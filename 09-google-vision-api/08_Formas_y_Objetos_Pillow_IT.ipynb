La Vision API te da números: riquadros delimitadores y vértices. Para *ver* lo que detectó, necesitas dibujar esos números sobre la immagine — y ahí entra **Pillow** (heredera de la histórica librería PIL, Python Imaging Library), la librería estándar de Python para abrir, dibujar sobre y guardar immagini.

````python
import os
# os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "ruta/a/tu/clave.json"

from google.cloud import vision
from PIL import Image, ImageDraw, ImageFont

client = vision.ImageAnnotatorClient()
print("Cliente listo.")
````

## El flujo general

1. La Vision API detecta oggetti y te devuelve coordenadas (`bounding_poly.vertices` o `normalized_vertices`)
2. Pillow abre la immagine y crea un oggetto de dibujo (`ImageDraw.Draw`) sobre ella
3. Pillow traza rectángulos, righe o polígonos usando esas coordenadas
4. Opcionalmente, Pillow añade una etichetta de testo junto a cada oggetto
5. Pillow guarda o muestra la immagine resultante

## Dibujar un riquadro delimitador simple

````python
def dibujar_bbox_simple(image_path, object_bbox, output_path):
    image = Image.open(image_path)
    draw = ImageDraw.Draw(image)
    draw.rectangle(object_bbox, outline="red", width=2)
    image.save(output_path)
    return output_path


# dibujar_bbox_simple("imagen.jpg", (50, 50, 200, 200), "imagen_anotada.jpg")
````

## Función `draw_vertices`: unir vértices y cerrar el polígono

Cuando trabajas con `object_localization` o `landmark_detection`, la Vision API te da una lista de vértices (no solo dos esquinas), así que conviene una funzione que una cada vértice con el siguiente **y cierre el contorno**.

````python
from io import BytesIO

def draw_vertices(image_source, vertices, display_text="", color="green", width=4):
    if isinstance(image_source, bytes):
        pillow_img = Image.open(BytesIO(image_source))
    else:
        pillow_img = Image.open(image_source)

    draw = ImageDraw.Draw(pillow_img)

    for i in range(len(vertices) - 1):
        draw.line(
            [(vertices[i][0], vertices[i][1]), (vertices[i + 1][0], vertices[i + 1][1])],
            fill=color,
            width=width,
        )

    # Línea de cierre: del último vértice al primero
    draw.line(
        [(vertices[-1][0], vertices[-1][1]), (vertices[0][0], vertices[0][1])],
        fill=color,
        width=width,
    )

    if display_text:
        font = ImageFont.load_default()
        draw.text((vertices[0][0] + 10, vertices[0][1]), display_text, font=font, fill="white")

    return pillow_img
````
