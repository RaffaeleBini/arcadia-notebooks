## 1. La analogía: los datos como ingredientes

Antes de cocinar necesitas saber:

- Qué ingredientes hacen falta (**contenido**)
- En qué formato los necesitas (**formato**)
- De dónde los vas a sacar (**fuente**)

> Igual que en la cocina, un ingrediente puede estar "fuera de temporada": más difícil de conseguir o más caro de lo previsto. Eso es exactamente lo que se revisa en la etapa de Data Collection.

## 2. Data Requirements — caso real: pacientes con CHF

Objetivo: definir el **contenido**, **formato** y **fuente** de datos para el árbol de decisión.

### El cohorte de pacientes: 3 criterios de inclusión

1. Admitido como paciente interno dentro del área de servicio de la aseguradora.
2. Diagnóstico principal de insuficiencia cardíaca congestiva (CHF) durante un año completo.
3. Al menos 6 meses de inscripción continua antes de la admisión principal por CHF.

**Se excluyó** a pacientes de CHF con otras condiciones médicas significativas: esas condiciones disparan tasas de reingreso más altas y distorsionarían el modelo.

### El formato exigido: un registro por paciente

- El árbol de decisión necesita **una fila por paciente**, con columnas = variables del modelo.
- El historial clínico real trae miles de registros transaccionales por paciente (admisiones, diagnósticos, procedimientos, prescripciones...).
- Hay que agregar (*rollup*) esos registros a nivel de paciente, creando nuevas variables resumen.
- Ese trabajo de agregación pertenece técnicamente a la siguiente etapa (Data Preparation), pero se anticipa aquí.

## 3. Data Collection — caso real: revisar y decidir

Tras la recolección inicial, toca preguntarse: **¿tengo lo que necesito?**

Fuentes usadas en el caso CHF:

- Información demográfica (edad, género, ubicación)
- Información clínica
- Información de cobertura del seguro
- Información del proveedor de salud
- Registros de reclamaciones (*claims*)
- Información farmacéutica y de otros diagnósticos

### El punto clave: está bien aplazar un dato

- Cierta información de **fármacos** no estaba integrada todavía con el resto de fuentes.
- Decisión: **aplazar** su recolección, no bloquear el proyecto.
- Se puede retomar más tarde, incluso después de tener **resultados intermedios** del modelo predictivo, si esos resultados sugieren que el dato importa.
- En este caso real: el equipo construyó un modelo razonablemente bueno **sin** esa información de fármacos.

> No todos los datos son igual de críticos. Saber cuáles puedes posponer es tan importante como saber cuáles necesitas ya.

Una vez con los datos recolectados, se aplica **estadística descriptiva** y **visualización** para evaluar contenido, calidad y detectar huecos.

## 4. Tres formas en las que llegan los datos

- **Estructurados**: tablas, filas y columnas bien definidas (hoja de cálculo, tabla de base de datos).
- **No estructurados**: sin formato predefinido — texto libre, imágenes, audio, vídeo. Requieren técnicas especiales.
- **Semiestructurados**: organización parcial (etiquetas, marcadores) pero sin la rigidez de una tabla — por ejemplo, recetas con nombre + lista variable de ingredientes.

Reconocer el tipo de dato te dice qué herramientas vas a necesitar.

## 5. Integración de datos: DBAs y programadores

Cuando los datos vienen de fuentes distintas (demográficos, clínicos, cobertura, reclamaciones...), alguien tiene que extraerlos y fusionarlos.

- Ese trabajo lo hacen los **DBA** (administradores de bases de datos) junto a programadores.
- Tarea: extraer, fusionar, y **eliminar redundancia**.
- El resultado queda listo para la siguiente etapa: Data Understanding.
- También es el momento de pensar en automatizar procesos de base de datos para futuras recolecciones.

## 7. Cargar el dataset con pandas

- **pandas**: la librería estándar de Python para trabajar con datos en tablas.
- **import**: instrucción que trae el código de una librería para poder usarla.
- **DataFrame**: el objeto central de pandas — una tabla en memoria, con filas y columnas, como una hoja de cálculo pero manipulable con código.

```python
import pandas as pd  # trae pandas y le pone el alias "pd", por convención

# Le pide a pandas que muestre todas las columnas al imprimir la tabla,
# en vez de recortar las columnas centrales con puntos suspensivos.
pd.set_option('display.max_columns', None)

# pd.read_csv() descarga el archivo CSV desde la URL y lo convierte
# directamente en un DataFrame: una tabla de filas y columnas en memoria.
recipes = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0103EN-SkillsNetwork/labs/Module%202/recipes.csv")

print("Data read into dataframe!")  # tarda unos 30 segundos en descargar
```
