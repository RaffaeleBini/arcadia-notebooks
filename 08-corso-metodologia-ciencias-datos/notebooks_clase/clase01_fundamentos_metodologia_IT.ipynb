## 1. Cos'è la scienza dei dati?

La scienza dei dati combina tre ingredienti:

- **Analisi statistica**
- **Conoscenza tecnologica** (programmazione, strumenti, infrastruttura)
- **Conoscenza del dominio** (il business da cui provengono i dati)

La combinazione dei tre è ciò che permette di estrarre informazioni preziose da grandi insiemi di dati.

> Avere molta potenza di calcolo e accesso facile ai dati **non basta**. Senza un processo chiaro, è facile non capire bene la domanda che ti stai ponendo, o non sapere applicare correttamente i dati al problema.

## 2. Cos'è una metodologia?

Una **metodologia** è un sistema di metodi usati in un'area di studio. È la guida delle decisioni da prendere durante un processo di ricerca.

Nella scienza dei dati, la metodologia:

- È un approccio strutturato per risolvere problemi complessi e prendere decisioni basate sui dati.
- Include modi di raccogliere dati, strategie di misurazione e confronto di metodi di analisi.
- Dà la guida pratica per fare ricerca in modo efficiente.

### La tentazione da evitare

Esiste la tentazione costante di **saltare direttamente alla soluzione** (aprire il notebook, addestrare un modello) senza passare dal processo. Quella tentazione sabota i tuoi stessi risultati.

**Analogia:** costruire una casa senza progetti. Puoi tirare su pareti il primo giorno, ma senza progetti finisci per buttare via lavoro fatto quando scopri che qualcosa non torna.

## 3. John Rollins e la metodologia a 10 fasi

- Proposta da **John Rollins**, data scientist senior in IBM.
- È **ciclica e iterativa**: non è una linea retta, torni indietro costantemente.
- Ha **10 fasi**, da *Business Understanding* a *Feedback*.

| # | Fase (inglese) | Fase (italiano) | A cosa risponde |
|---|---|---|---|
| 1 | Business Understanding | Comprensione del business | Qual è il problema reale da risolvere? |
| 2 | Analytic Approach | Approccio analitico | Che analisi o tecnica serve? |
| 3 | Data Requirements | Requisiti dei dati | Quali dati concreti mi servono? |
| 4 | Data Collection | Raccolta dei dati | Da dove vengono e come li ottengo? |
| 5 | Data Understanding | Comprensione dei dati | I dati rappresentano il problema? |
| 6 | Data Preparation | Preparazione dei dati | Cosa va pulito o trasformato? |
| 7 | Modeling | Modellazione | Che modello costruisco? |
| 8 | Evaluation | Valutazione | Il modello risponde alla domanda di business? |
| 9 | Deployment | Distribuzione | Posso portare il modello in pratica? |
| 10 | Feedback | Retroazione | Che feedback danno i dati e le persone interessate? |

## 4. Le 10 domande chiave, in 4 blocchi

**Blocco 1 — Definire il problema e l'approccio** (fasi 1-2)
1. Qual è il problema che stai cercando di risolvere?
2. Come puoi usare i dati per rispondere a quella domanda?

**Blocco 2 — Organizzarti attorno ai dati** (fasi 3-6)
3. Quali dati ti servono?
4. Da dove viene la fonte dei dati e come la ricevi?
5. I dati raccolti rappresentano il problema da risolvere?
6. Che lavoro aggiuntivo serve per manipolarli?

**Blocco 3 — Modellare** (fasi 7-8)
7. Le visualizzazioni mostrano risposte che affrontano il problema di business?
8. Il modello risponde alla domanda di business iniziale, o bisogna aggiustare i dati?

**Blocco 4 — Validare l'approccio e il design finale** (fasi 9-10)
9. Puoi mettere il modello in pratica?
10. Puoi ottenere feedback costruttivo dai dati e dalle persone interessate?

```python
# Mappatura semplice: blocco di domande -> fasi di Rollins che copre.
# Anche questo non è materiale ufficiale, è solo supporto visivo per la lezione.

bloques_preguntas = {
    "1. Definire problema e approccio": ["1. Business Understanding", "2. Analytic Approach"],
    "2. Organizzarti con i dati": [
        "3. Data Requirements", "4. Data Collection",
        "5. Data Understanding", "6. Data Preparation",
    ],
    "3. Modellare": ["7. Modeling", "8. Evaluation"],
    "4. Validare approccio e design finale": ["9. Deployment", "10. Feedback"],
}

for bloque, etapas in bloques_preguntas.items():
    print(bloque)
    for etapa in etapas:
        print(f"   - {etapa}")
```

## 5. CRISP-DM: l'altra metodologia di riferimento

**CRISP-DM** = Cross-Industry Standard Process for Data Mining.

- Modello iterativo e ciclico, collaudato nell'industria per il data mining.
- Richiede flessibilità e comunicazione costante con le persone interessate.
- Ha **6 fasi**:

1. **Business Understanding** — la fase più importante; fissa le intenzioni del progetto.
2. **Data Understanding** — si decide sulle fonti di dati e si acquisiscono.
3. **Data Preparation** — si trasformano i dati in un sottoinsieme utilizzabile.
4. **Modeling** — si creano modelli che rivelano pattern ("un'arte e una scienza").
5. **Evaluation** — si testa il modello con dati trattati come nuovi.
6. **Deployment** — si usa il modello con dati nuovi, fuori dal dataset originale.

### La differenza chiave

CRISP-DM **fonde** le fasi di Rollins *Data Requirements*, *Data Collection* e *Data Understanding* **in un'unica fase** chiamata Data Understanding.

Dopo le 6 fasi c'è un'altra riunione di Business Understanding con le persone interessate — ma quel ritorno **non ha nome proprio** in CRISP-DM. In Rollins invece sì: è la fase 10, **Feedback**.

### Confronto fianco a fianco

| Rollins (10 fasi) | CRISP-DM (6 fasi) |
|---|---|
| 1. Business Understanding | 1. Business Understanding |
| 2. Analytic Approach | *(nessuna fase esplicita)* |
| 3. Data Requirements | 2. Data Understanding |
| 4. Data Collection | 2. Data Understanding |
| 5. Data Understanding | 2. Data Understanding |
| 6. Data Preparation | 3. Data Preparation |
| 7. Modeling | 4. Modeling |
| 8. Evaluation | 5. Evaluation |
| 9. Deployment | 6. Deployment |
| 10. Feedback | *(avviene comunque, senza nome proprio)* |

```python
# Struttura dati che rappresenta la fusione delle fasi di Rollins
# dentro le fasi di CRISP-DM. Esempio proprio per la lezione, non ufficiale.

fases_crispdm_a_rollins = {
    "1. Business Understanding": ["1. Business Understanding"],
    "2. Data Understanding": ["3. Data Requirements", "4. Data Collection", "5. Data Understanding"],
    "3. Data Preparation": ["6. Data Preparation"],
    "4. Modeling": ["7. Modeling"],
    "5. Evaluation": ["8. Evaluation"],
    "6. Deployment": ["9. Deployment"],
}

print("Come si raggruppano le 10 fasi di Rollins dentro le 6 fasi di CRISP-DM:\n")
for fase_crispdm, etapas_rollins in fases_crispdm_a_rollins.items():
    etiqueta = "1 fase" if len(etapas_rollins) == 1 else f"{len(etapas_rollins)} fasi fuse"
    print(f"{fase_crispdm}  ({etiqueta})")
    for etapa in etapas_rollins:
        print(f"   - {etapa}")

# Nota: Analytic Approach (2) e Feedback (10) di Rollins non hanno
# una fase esplicita equivalente in CRISP-DM.
etapas_rollins_todas = {e for etapa, pregunta in metodologia_rollins if (e := etapa)}
etapas_mapeadas = {e for etapas in fases_crispdm_a_rollins.values() for e in etapas}
sin_fase_explicita = etapas_rollins_todas - etapas_mapeadas
print("\nFasi di Rollins senza fase esplicita in CRISP-DM:", sorted(sin_fase_explicita))
```

## Riepilogo della lezione

- La scienza dei dati = statistica + tecnologia + conoscenza del business.
- Una metodologia evita che tu salti direttamente a "la soluzione" senza capire il problema.
- Rollins (IBM): 10 fasi, cicliche e iterative, da Business Understanding a Feedback.
- Le 10 fasi rispondono a 10 domande raggruppate in 4 blocchi.
- CRISP-DM: 6 fasi, fonde 3 fasi di Rollins in una sola (Data Understanding).
- Il "Feedback" di Rollins avviene anche in CRISP-DM, ma senza nome proprio.
