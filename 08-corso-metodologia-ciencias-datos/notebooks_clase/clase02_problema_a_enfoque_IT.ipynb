## 1. Perché il business viene prima dei dati

- La metodologia di Rollins inizia con **Business Understanding**, non con un dataset.
- Analogia: il tuo capo ti dà un compito urgente in una riunione → quel pomeriggio scopri che ti mancano domande da fare → il tuo capo non è disponibile fino a domani → vai avanti alla cieca o ti fermi a chiarire?
- Avere chiarezza sul problema è ciò che ti permette di determinare **quali dati** ti serviranno.
- Errore più comune: risolvere molto bene ciò che la gente CREDE sia la domanda, invece della domanda reale.

## 2. Goal vs. Objectives

- **Goal** = l'obiettivo ultimo di chi chiede.
- **Objectives** = i passi concreti e misurabili che sostengono quell'obiettivo.

Esempio dalla trascrizione di IBM:

> "Come possiamo ridurre il costo di svolgere un'attività?"

Questa domanda può nascondere due goal diversi:

- Migliorare l'**efficienza** dell'attività.
- Aumentare la **redditività** del business.

Ogni goal porta a objective e a un'analisi diversa. Chiarire il goal è il primo passo, sempre.

Lo **sponsor del progetto** (business sponsor) ha un ruolo chiave:
- Segna la direzione generale.
- Resta coinvolto dando orientamento.
- Garantisce il supporto necessario quando serve.

## 3. Il caso reale: ricoveri ospedalieri ripetuti per insufficienza cardiaca congestizia (CHF)

Caso reale di IBM. **Non ha codice eseguibile**: avviene completamente prima di toccare dati.

**Contesto di business:**
- Un'assicurazione sanitaria statunitense vede scendere il finanziamento pubblico per i ricoveri ripetuti.
- Rischio di dover aumentare le tariffe ai propri clienti se non agisce.
- L'assicurazione si unisce alle autorità sanitarie e porta data scientist di IBM.

**Domanda centrale:** qual è il modo migliore di assegnare un budget sanitario limitato, per massimizzarne l'uso nel dare un'assistenza di qualità?

**Scoperte prima di avere il modello:**
- ~30% di pazienti riammessi in un centro di riabilitazione entro 1 anno.
- ~50% riammessi entro 5 anni.
- I pazienti con **insufficienza cardiaca congestizia (CHF)** sono in cima alla lista dei ricoveri ripetuti.

**Processo:** IBM ha proposto e realizzato un workshop in presenza per fissare goal e objective con lo sponsor di business.

**4 requisiti di business identificati nel workshop:**
1. Prevedere i risultati di ricovero per pazienti con CHF.
2. Prevedere il rischio di ricovero.
3. Capire la combinazione di eventi che ha portato al risultato previsto.
4. Applicare un processo facile da capire per valutare nuovi pazienti.

## 4. I cinque approcci analitici

| Approccio | A cosa risponde |
|---|---|
| **Descrittivo** | Stato attuale / relazioni (es. cluster di attività simili) |
| **Diagnostico** | Analisi statistica (contare, spiegare perché è successo qualcosa) |
| **Predittivo** | Previsione / probabilità di un'azione futura |
| **Prescrittivo** | Raccomandazioni concrete su cosa fare |
| **Machine Learning (ML)** | Imparare pattern senza programmazione esplicita; clustering/associazione per il comportamento umano |

## 5. Il tipo di domanda determina l'approccio

- Domanda di **probabilità** → approccio **predittivo**.
- Domanda di **relazioni / cose simili** → approccio **descrittivo** (o clustering).
- Domanda di **sì o no** → approccio di **classificazione**.
- Domanda sul **comportamento umano** → **clustering / associazione**.

Questa mappatura è ciò che ti porta da "ho una domanda di business" a "so che tipo di modello mi serve", prima di guardare un solo dato.

## 6. Perché si è scelto un albero decisionale per il caso CHF

Si è usato un modello di **classificazione ad albero decisionale** (decision tree classification):

- Esamina variabili in ogni nodo, lungo ogni percorso fino a una foglia, fino a un valore soglia.
- Dà la **previsione** (riammissione sì/no) e la **probabilità** di quel risultato contemporaneamente (proporzione del risultato dominante in ogni foglia).

**Perché si adattava ai 4 requisiti di business:**
1. Facile da capire per chi non è un data scientist.
2. Dà previsione + probabilità contemporaneamente.
3. I clinici vedono esattamente quali condizioni portano un paziente a essere ad alto rischio — permette di costruire più modelli in diversi momenti della degenza ospedaliera, dando un'immagine in movimento del rischio.

Con questo, la classificazione ad albero decisionale è stato l'approccio scelto per il modello di ricovero per CHF.

## 7. Un esempio di codice per fissare l'idea

Questo esempio **non viene da nessun notebook ufficiale di IBM**. Serve solo a esercitarti, con Python puro (senza librerie esterne), sulla logica di mappatura domanda → approccio analitico appena vista.

```python
# Questo codice NON è materiale ufficiale di IBM. È un esempio mio,
# pensato unicamente per esercitarsi sull'idea di "il tipo di domanda
# determina l'approccio analitico" con qualcosa di eseguibile.

def sugerir_enfoque_analitico(pregunta):
    """
    Riceve una domanda di business come testo e suggerisce un approccio
    analitico, cercando parole chiave tipiche di ogni tipo di domanda.
    """
    pregunta = pregunta.lower()

    if "probabilidad" in pregunta or "cuántos" in pregunta or "cuantos" in pregunta:
        return "Predictivo (pronóstico de una probabilidad o cantidad futura)"
    elif "relación" in pregunta or "relacion" in pregunta or "parecidos" in pregunta or "similares" in pregunta:
        return "Descriptivo (mostrar relaciones o agrupar actividades similares)"
    elif "sí o no" in pregunta or "si o no" in pregunta:
        return "Clasificación (una respuesta binaria: sí/no)"
    elif "comportamiento" in pregunta:
        return "Machine Learning: Clustering / Asociación (entender comportamiento humano)"
    else:
        return "No se detectaron palabras clave claras: aclara la pregunta con el stakeholder"

print("Función definida correctamente.")
```

### Proviamo la funzione con domande di esempio

```python
preguntas_de_ejemplo = [
    "¿Cuál es la probabilidad de que este paciente sea readmitido en 30 días?",
    "¿Qué pacientes tienen hábitos de consumo parecidos entre sí?",
    "¿Este paciente va a ser readmitido, sí o no?",
    "¿Qué patrones de comportamiento explican por qué un paciente no sigue el tratamiento?",
]

for pregunta in preguntas_de_ejemplo:
    enfoque = sugerir_enfoque_analitico(pregunta)
    print(f"Pregunta: {pregunta}")
    print(f"  -> Enfoque sugerido: {enfoque}")
    print()
```
