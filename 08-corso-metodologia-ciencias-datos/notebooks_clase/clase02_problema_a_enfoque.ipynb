## 1. Por qué el negocio va antes que los datos

- La metodología de Rollins empieza con **Business Understanding**, no con un dataset.
- Analogía: tu jefe te da una tarea urgente en una reunión → esa tarde descubres que te faltan preguntas por hacer → tu jefe no está disponible hasta mañana → ¿avanzas a ciegas o te detienes a aclarar?
- Tener claridad sobre el problema es lo que te permite determinar **qué datos** vas a necesitar.
- Error más común: resolver muy bien lo que la gente CREE que es la pregunta, en vez de la pregunta real.

## 2. Goal vs. Objectives

- **Goal** = la meta última de quien pregunta.
- **Objectives** = los pasos concretos y medibles que apoyan esa meta.

Ejemplo de la transcripción de IBM:

> "¿Cómo podemos reducir el coste de realizar una actividad?"

Esta pregunta puede esconder dos goals distintos:

- Mejorar la **eficiencia** de la actividad.
- Aumentar la **rentabilidad** del negocio.

Cada goal lleva a objectives y a un análisis distinto. Aclarar el goal es el primer paso, siempre.

El **patrocinador del proyecto** (business sponsor) tiene un rol clave:
- Marca la dirección general.
- Se mantiene involucrado dando orientación.
- Asegura el apoyo necesario cuando hace falta.

## 3. El caso real: reingresos hospitalarios por insuficiencia cardíaca congestiva (CHF)

Caso real de IBM. **No tiene código ejecutable**: ocurre completamente antes de tocar datos.

**Contexto de negocio:**
- Una aseguradora de salud estadounidense ve caer la financiación pública para reingresos.
- Riesgo de tener que subir tarifas a sus clientes si no actúa.
- La aseguradora se junta con autoridades de salud y trae científicos de datos de IBM.

**Pregunta central:** ¿cuál es la mejor forma de asignar un presupuesto de salud limitado, para maximizar su uso en dar atención de calidad?

**Hallazgos antes de tener el modelo:**
- ~30% de pacientes readmitidos en un centro de rehabilitación dentro de 1 año.
- ~50% readmitidos dentro de 5 años.
- Los pacientes con **insuficiencia cardíaca congestiva (CHF)** encabezan la lista de reingresos.

**Proceso:** IBM propuso y realizó un workshop presencial para fijar goals y objectives con el patrocinador del negocio.

**4 requisitos de negocio identificados en el workshop:**
1. Predecir los resultados de reingreso para pacientes con CHF.
2. Predecir el riesgo de reingreso.
3. Entender la combinación de eventos que llevó al resultado predicho.
4. Aplicar un proceso fácil de entender para evaluar a nuevos pacientes.

## 4. Los cinco enfoques analíticos

| Enfoque | Qué responde |
|---|---|
| **Descriptivo** | Estado actual / relaciones (ej. clusters de actividades similares) |
| **Diagnóstico** | Análisis estadístico (contar, explicar por qué pasó algo) |
| **Predictivo** | Pronóstico / probabilidad de una acción futura |
| **Prescriptivo** | Recomendaciones concretas sobre qué hacer |
| **Machine Learning (ML)** | Aprender patrones sin programación explícita; clustering/asociación para comportamiento humano |

## 5. El tipo de pregunta determina el enfoque

- Pregunta de **probabilidad** → enfoque **predictivo**.
- Pregunta de **relaciones / cosas parecidas** → enfoque **descriptivo** (o clustering).
- Pregunta de **sí o no** → enfoque de **clasificación**.
- Pregunta sobre **comportamiento humano** → **clustering / asociación**.

Este mapeo es lo que te lleva de "tengo una pregunta de negocio" a "sé qué tipo de modelo necesito", antes de mirar un solo dato.

## 6. Por qué se eligió un árbol de decisión para el caso CHF

Se usó un modelo de **clasificación por árbol de decisión** (decision tree classification):

- Examina variables en cada nodo, a lo largo de cada camino hasta una hoja, hasta un valor umbral.
- Da la **predicción** (readmisión sí/no) y la **probabilidad** de ese resultado a la vez (proporción del resultado dominante en cada hoja).

**Por qué encajaba con los 4 requisitos de negocio:**
1. Fácil de entender para quien no es científico de datos.
2. Da predicción + probabilidad al mismo tiempo.
3. Los clínicos ven exactamente qué condiciones llevan a un paciente a ser de alto riesgo — permite construir varios modelos en distintos momentos de la estancia hospitalaria, dando una imagen en movimiento del riesgo.

Con esto, la clasificación por árbol de decisión fue el enfoque elegido para el modelo de reingreso por CHF.

## 7. Un ejemplo de código para fijar la idea

Este ejemplo **no viene de ningún notebook oficial de IBM**. Es solo para practicar, con Python puro (sin librerías externas), la lógica de mapeo pregunta → enfoque analítico que acabas de ver.

```python
# Este código NO es material oficial de IBM. Es un ejemplo propio,
# pensado únicamente para practicar la idea de "el tipo de pregunta
# determina el enfoque analítico" con algo ejecutable.

def sugerir_enfoque_analitico(pregunta):
    """
    Recibe una pregunta de negocio como texto y sugiere un enfoque
    analítico, buscando palabras clave típicas de cada tipo de pregunta.
    """
    pregunta = pregunta.lower()

    if "probabilidad" in pregunta or "cuántos" in pregunta or "cuantos" in pregunta:
        return "Predictivo (pronóstico de una probabilidad o cantidad futura)"
    elif "relación" in pregunta or "relacion" in pregunta or "parecidos" in pregunta or "similares" in pregunta:
        return "Descriptivo (mostrar relaciones o agrupar actividades similares)"
    elif "sí o no" in pregunta or "si o no" in pregunta:
        return "Clasificación (una respuesta binaria: sí/no)"
    elif "comportamiento" in pregunta:
        return "Machine Learning: Clustering / Asociación (entender comportamiento humano)"
    else:
        return "No se detectaron palabras clave claras: aclara la pregunta con el stakeholder"

print("Función definida correctamente.")
```

### Probemos la función con preguntas de ejemplo

```python
preguntas_de_ejemplo = [
    "¿Cuál es la probabilidad de que este paciente sea readmitido en 30 días?",
    "¿Qué pacientes tienen hábitos de consumo parecidos entre sí?",
    "¿Este paciente va a ser readmitido, sí o no?",
    "¿Qué patrones de comportamiento explican por qué un paciente no sigue el tratamiento?",
]

for pregunta in preguntas_de_ejemplo:
    enfoque = sugerir_enfoque_analitico(pregunta)
    print(f"Pregunta: {pregunta}")
    print(f"  -> Enfoque sugerido: {enfoque}")
    print()
```

## Resumen de la clase

- **Business Understanding** siempre va antes que los datos: define el problema real primero.
- **Goal** = meta última; **objectives** = pasos concretos que la apoyan. Una misma pregunta puede esconder metas distintas.
- El **patrocinador del proyecto** marca dirección, se mantiene involucrado y asegura el apoyo necesario.
- Caso CHF: aseguradora de salud + IBM, 30%/50% de reingresos, workshop, 4 requisitos de negocio.
- 5 enfoques analíticos: descriptivo, diagnóstico, predictivo, prescriptivo, machine learning.
- El tipo de pregunta determina el enfoque: probabilidad → predictivo; relaciones → descriptivo; sí/no → clasificación; comportamiento → clustering/asociación.
- Se eligió árbol de decisión para CHF: fácil de interpretar, da predicción + probabilidad, y muestra qué condiciones llevan al riesgo.
