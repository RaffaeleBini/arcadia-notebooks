## 1. ¿Qué es la ciencia de datos?

La ciencia de datos combina tres ingredientes:

- **Análisis estadístico**
- **Conocimiento tecnológico** (programación, herramientas, infraestructura)
- **Conocimiento del dominio** (el negocio del que salen los datos)

La combinación de los tres es lo que permite extraer información valiosa de grandes conjuntos de datos.

> Tener mucha potencia de cálculo y acceso fácil a los datos **no es suficiente**. Sin un proceso claro, es fácil no entender bien la pregunta que te estás haciendo, o no saber aplicar los datos correctamente al problema.

## 2. ¿Qué es una metodología?

Una **metodología** es un sistema de métodos usados en un área de estudio. Es la guía de las decisiones que hay que tomar durante un proceso de investigación.

En ciencia de datos, la metodología:

- Es un enfoque estructurado para resolver problemas complejos y tomar decisiones basadas en datos.
- Incluye formas de recolectar datos, estrategias de medición y comparación de métodos de análisis.
- Da la guía práctica para investigar de forma eficiente.

### La tentación que hay que evitar

Existe la tentación constante de **saltar directo a la solución** (abrir el notebook, entrenar un modelo) sin pasar por el proceso. Esa tentación sabotea tus propios resultados.

**Analogía:** construir una casa sin planos. Puedes levantar paredes el primer día, pero sin planos acabas tirando trabajo hecho cuando descubres que algo no encaja.

## 3. John Rollins y la metodología de 10 etapas

- Propuesta por **John Rollins**, científico de datos senior en IBM.
- Es **cíclica e iterativa**: no es una línea recta, vuelves atrás constantemente.
- Tiene **10 etapas**, de *Business Understanding* a *Feedback*.

| # | Etapa (inglés) | Etapa (español) | Qué responde |
|---|---|---|---|
| 1 | Business Understanding | Comprensión del negocio | ¿Cuál es el problema real que hay que resolver? |
| 2 | Analytic Approach | Enfoque analítico | ¿Qué análisis o técnica hace falta? |
| 3 | Data Requirements | Requisitos de datos | ¿Qué datos concretos necesito? |
| 4 | Data Collection | Recolección de datos | ¿De dónde vienen y cómo los consigo? |
| 5 | Data Understanding | Comprensión de los datos | ¿Los datos representan el problema? |
| 6 | Data Preparation | Preparación de datos | ¿Qué hay que limpiar o transformar? |
| 7 | Modeling | Modelado | ¿Qué modelo construyo? |
| 8 | Evaluation | Evaluación | ¿El modelo responde la pregunta de negocio? |
| 9 | Deployment | Despliegue | ¿Puedo llevar el modelo a la práctica? |
| 10 | Feedback | Retroalimentación | ¿Qué feedback dan los datos y los interesados? |

```python
# Este código NO es material oficial de IBM: es solo una forma de
# visualizar las 10 etapas de Rollins como estructura de datos en Python.

metodologia_rollins = [
    ("1. Business Understanding", "¿Cuál es el problema que hay que resolver?"),
    ("2. Analytic Approach", "¿Cómo uso datos para responder la pregunta?"),
    ("3. Data Requirements", "¿Qué datos necesito?"),
    ("4. Data Collection", "¿De dónde vienen y cómo los recibo?"),
    ("5. Data Understanding", "¿Los datos representan el problema?"),
    ("6. Data Preparation", "¿Qué hay que limpiar o transformar?"),
    ("7. Modeling", "¿Qué modelo construyo?"),
    ("8. Evaluation", "¿El modelo responde la pregunta de negocio?"),
    ("9. Deployment", "¿Puedo poner el modelo en práctica?"),
    ("10. Feedback", "¿Qué feedback dan los datos y los interesados?"),
]

for etapa, pregunta in metodologia_rollins:
    print(f"{etapa:30s} -> {pregunta}")
```

## 4. Las 10 preguntas clave, en 4 bloques

**Bloque 1 — Definir el problema y el enfoque** (etapas 1-2)
1. ¿Cuál es el problema que intentas resolver?
2. ¿Cómo puedes usar datos para responder esa pregunta?

**Bloque 2 — Organizarte alrededor de los datos** (etapas 3-6)
3. ¿Qué datos necesitas?
4. ¿De dónde sale la fuente de datos y cómo la recibes?
5. ¿Los datos recolectados representan el problema a resolver?
6. ¿Qué trabajo adicional hace falta para manipularlos?

**Bloque 3 — Modelar** (etapas 7-8)
7. ¿Las visualizaciones muestran respuestas que atacan el problema de negocio?
8. ¿El modelo responde la pregunta de negocio inicial, o hay que ajustar los datos?

**Bloque 4 — Validar el enfoque y el diseño final** (etapas 9-10)
9. ¿Puedes poner el modelo en práctica?
10. ¿Puedes conseguir feedback constructivo de los datos y los interesados?

```python
# Mapeo simple: bloque de preguntas -> etapas de Rollins que cubre.
# Tampoco es material oficial, es solo apoyo visual para la clase.

bloques_preguntas = {
    "1. Definir problema y enfoque": ["1. Business Understanding", "2. Analytic Approach"],
    "2. Organizarte con los datos": [
        "3. Data Requirements", "4. Data Collection",
        "5. Data Understanding", "6. Data Preparation",
    ],
    "3. Modelar": ["7. Modeling", "8. Evaluation"],
    "4. Validar enfoque y diseño final": ["9. Deployment", "10. Feedback"],
}

for bloque, etapas in bloques_preguntas.items():
    print(bloque)
    for etapa in etapas:
        print(f"   - {etapa}")
```

## 5. CRISP-DM: la otra metodología de referencia

**CRISP-DM** = Cross-Industry Standard Process for Data Mining.

- Modelo iterativo y cíclico, probado en la industria para minería de datos.
- Requiere flexibilidad y comunicación constante con los interesados.
- Tiene **6 fases**:

1. **Business Understanding** — la fase más importante; fija las intenciones del proyecto.
2. **Data Understanding** — se decide sobre fuentes de datos y se adquieren.
3. **Data Preparation** — se transforman los datos en un subconjunto usable.
4. **Modeling** — se crean modelos que revelan patrones ("un arte y una ciencia").
5. **Evaluation** — se prueba el modelo con datos tratados como nuevos.
6. **Deployment** — se usa el modelo con datos nuevos, fuera del dataset original.

### La diferencia clave

CRISP-DM **fusiona** las etapas de Rollins *Data Requirements*, *Data Collection* y *Data Understanding* **en una sola fase** llamada Data Understanding.

Después de las 6 fases hay otra reunión de Business Understanding con los interesados — pero esa vuelta **no tiene nombre propio** en CRISP-DM. En Rollins sí lo tiene: es la etapa 10, **Feedback**.

### Comparación lado a lado

| Rollins (10 etapas) | CRISP-DM (6 fases) |
|---|---|
| 1. Business Understanding | 1. Business Understanding |
| 2. Analytic Approach | *(sin fase explícita)* |
| 3. Data Requirements | 2. Data Understanding |
| 4. Data Collection | 2. Data Understanding |
| 5. Data Understanding | 2. Data Understanding |
| 6. Data Preparation | 3. Data Preparation |
| 7. Modeling | 4. Modeling |
| 8. Evaluation | 5. Evaluation |
| 9. Deployment | 6. Deployment |
| 10. Feedback | *(ocurre igual, sin nombre propio)* |

```python
# Estructura de datos que representa la fusión de etapas de Rollins
# dentro de las fases de CRISP-DM. Ejemplo propio para la clase, no oficial.

fases_crispdm_a_rollins = {
    "1. Business Understanding": ["1. Business Understanding"],
    "2. Data Understanding": ["3. Data Requirements", "4. Data Collection", "5. Data Understanding"],
    "3. Data Preparation": ["6. Data Preparation"],
    "4. Modeling": ["7. Modeling"],
    "5. Evaluation": ["8. Evaluation"],
    "6. Deployment": ["9. Deployment"],
}

print("Cómo se agrupan las 10 etapas de Rollins dentro de las 6 fases de CRISP-DM:\n")
for fase_crispdm, etapas_rollins in fases_crispdm_a_rollins.items():
    etiqueta = "1 etapa" if len(etapas_rollins) == 1 else f"{len(etapas_rollins)} etapas fusionadas"
    print(f"{fase_crispdm}  ({etiqueta})")
    for etapa in etapas_rollins:
        print(f"   - {etapa}")

# Nota: Analytic Approach (2) y Feedback (10) de Rollins no tienen
# una fase explícita equivalente en CRISP-DM.
etapas_rollins_todas = {e for etapa, pregunta in metodologia_rollins if (e := etapa)}
etapas_mapeadas = {e for etapas in fases_crispdm_a_rollins.values() for e in etapas}
sin_fase_explicita = etapas_rollins_todas - etapas_mapeadas
print("\nEtapas de Rollins sin fase explícita en CRISP-DM:", sorted(sin_fase_explicita))
```

## Resumen de la clase

- La ciencia de datos = estadística + tecnología + conocimiento del negocio.
- Una metodología evita que saltes directo a "la solución" sin entender el problema.
- Rollins (IBM): 10 etapas, cíclicas e iterativas, de Business Understanding a Feedback.
- Las 10 etapas responden 10 preguntas agrupadas en 4 bloques.
- CRISP-DM: 6 fases, funde 3 etapas de Rollins en una sola (Data Understanding).
- El "Feedback" de Rollins ocurre en CRISP-DM también, pero sin nombre propio.
