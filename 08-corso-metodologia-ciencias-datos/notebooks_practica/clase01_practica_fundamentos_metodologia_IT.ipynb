## Esercizio 1 — Riordina le 10 fasi

Ti do la lista delle 10 fasi di Rollins, ma **disordinata**. Il tuo compito è scrivere codice che le riordini secondo la sequenza corretta della metodologia (da Business Understanding a Feedback).

**Suggerimenti:**
- Ogni fase in `etapas_desordenadas` è una tupla `(numero, nombre)`, ad esempio `(7, "Modeling")`.
- Il numero dentro la tupla ti dice già la posizione corretta — non serve che tu la memorizzi di nuovo.
- Pensa a quale funzione di ordinamento di Python ti serve per ordinare una lista di tuple per il primo elemento.

````python
etapas_desordenadas = [
    (7, "Modeling"),
    (1, "Business Understanding"),
    (9, "Deployment"),
    (3, "Data Requirements"),
    (10, "Feedback"),
    (5, "Data Understanding"),
    (2, "Analytic Approach"),
    (8, "Evaluation"),
    (4, "Data Collection"),
    (6, "Data Preparation"),
]

# TODO: crea una nuova lista chiamata `etapas_ordenadas` che contenga le
# stesse tuple di `etapas_desordenadas` ma ordinate per il loro numero
# (da 1 a 10). Non modificare `etapas_desordenadas` direttamente.
etapas_ordenadas = None

# TODO: scorri `etapas_ordenadas` con un ciclo for e stampa ogni fase
# nel formato: "1. Business Understanding"
````

<details>
<summary>💡 Vedi soluzione</summary>

```python
etapas_desordenadas = [
    (7, "Modeling"),
    (1, "Business Understanding"),
    (9, "Deployment"),
    (3, "Data Requirements"),
    (10, "Feedback"),
    (5, "Data Understanding"),
    (2, "Analytic Approach"),
    (8, "Evaluation"),
    (4, "Data Collection"),
    (6, "Data Preparation"),
]

# sorted() con key=lambda etapa: etapa[0] ordina per il primo
# elemento di ogni tupla (il numero della fase), senza toccare la lista originale.
etapas_ordenadas = sorted(etapas_desordenadas, key=lambda etapa: etapa[0])

for numero, nombre in etapas_ordenadas:
    print(f"{numero}. {nombre}")
```
