primeras_columnas = recipes.columns[:10]
print(primeras_columnas)

# .iloc[:, 0] seleziona la prima colonna per posizione (indice 0),
# senza bisogno di conoscere il suo nome esatto. .nunique() conta quanti
# valori distinti appaiono in essa.
n_cocinas = recipes.iloc[:, 0].nunique()
print(f"Numero di cucine distinte: {n_cocinas}")
