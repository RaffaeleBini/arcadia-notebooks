## Ejercicio 1 — Reordena las 10 etapas

Te doy la lista de las 10 etapas de Rollins, pero **desordenada**. Tu tarea es escribir código que las reordene según la secuencia correcta de la metodología (de Business Understanding a Feedback).

**Pistas:**

<details>
<summary>💡 Ver solución</summary>

```python
etapas_desordenadas = [
    (7, "Modeling"),
    (1, "Business Understanding"),
    (9, "Deployment"),
    (3, "Data Requirements"),
    (10, "Feedback"),
    (5, "Data Understanding"),
    (2, "Analytic Approach"),
    (8, "Evaluation"),
    (4, "Data Collection"),
    (6, "Data Preparation"),
]

# sorted() con key=lambda etapa: etapa[0] ordena por el primer
# elemento de cada tupla (el número de la etapa), sin tocar la lista original.
etapas_ordenadas = sorted(etapas_desordenadas, key=lambda etapa: etapa[0])

for numero, nombre in etapas_ordenadas:
    print(f"{numero}. {nombre}")
```
