def top_n_ingredients(cuisine_name, n=3):
    """Restituisce una lista di (ingrediente, percentuale) con gli n ingredienti
    più frequenti della cucina 'cuisine_name', ordinati dal maggiore al minore."""
    # 1. riga della cucina scelta (Series: indice = ingrediente, valore = proporzione 0-1)
    row = cuisines.loc[cuisine_name]

    # 2. ordinare dal maggiore al minore e passare a percentuale
    row_sorted = row.sort_values(ascending=False) * 100

    # 3. tenere i primi n
    top_ingredients = list(row_sorted.index.values)[0:n]
    top_percentages = list(row_sorted.values)[0:n]

    # 4. combinare in tuple (ingrediente, percentuale)
    resultado = list(zip(top_ingredients, top_percentages))

    return resultado

# 1. filtrare ricette con le tre condizioni contemporaneamente
combo_recipes = recipes.loc[
    (recipes["garlic"] == 1) &
    (recipes["onion"] == 1) &
    (recipes["tomato"] == 1)
]

# 2. contare quante ce ne sono

print(f"Ricette con garlic + onion + tomato: {num_combo}")

# 3. top 5 cucine più frequenti in quel sottoinsieme
