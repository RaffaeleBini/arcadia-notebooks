def top_n_ingredients(cuisine_name, n=3):
    """Devuelve una lista de (ingrediente, porcentaje) con los n ingredientes
    más frecuentes de la cocina 'cuisine_name', ordenados de mayor a menor."""
    # 1. fila de la cocina elegida (Series: índice = ingrediente, valor = proporción 0-1)
    row = cuisines.loc[cuisine_name]

    # 2. ordenar de mayor a menor y pasar a porcentaje
    row_sorted = row.sort_values(ascending=False) * 100

    # 3. quedarnos con los primeros n
    top_ingredients = list(row_sorted.index.values)[0:n]
    top_percentages = list(row_sorted.values)[0:n]

    # 4. combinar en tuplas (ingrediente, porcentaje)
    resultado = list(zip(top_ingredients, top_percentages))

    return resultado
