## 0. Preprocesamiento (ya resuelto)

Esto ya lo hiciste en clases anteriores. Aquí lo dejamos ya resuelto para que puedas concentrarte en la parte de storytelling: la limpieza del dataset y el cálculo del perfil de ingredientes por cocina (`cuisines = recipes.groupby("cuisine").mean()`).

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

recipes = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DS0103EN-SkillsNetwork/labs/Module%202/recipes.csv")

column_names = recipes.columns.values
column_names[0] = "cuisine"
recipes.columns = column_names
recipes["cuisine"] = recipes["cuisine"].str.lower()

recipes_counts = recipes["cuisine"].value_counts()
cuisines_indices = recipes_counts > 50
cuisines_to_keep = list(np.array(recipes_counts.index.values)[np.array(cuisines_indices)])
recipes = recipes.loc[recipes["cuisine"].isin(cuisines_to_keep)]

recipes = recipes.replace(to_replace="Yes", value=1)
recipes = recipes.replace(to_replace="No", value=0)

# Perfil de ingredientes por cocina: el promedio de cada ingrediente
# dentro de cada cocina es el porcentaje de recetas de esa cocina
# que lo incluyen (visto en la Clase 4 / capítulo 4 del manual).
cuisines = recipes.groupby("cuisine").mean(numeric_only=True)

print("Cocinas disponibles:")
print(sorted(cuisines.index.tolist()))
````
