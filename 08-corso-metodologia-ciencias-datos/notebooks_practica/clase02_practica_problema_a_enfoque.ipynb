def sugerir_enfoque_analitico(pregunta):
    pregunta = pregunta.lower()
    if "probabilidad" in pregunta or "cuántos" in pregunta or "cuantos" in pregunta:
        return "Predictivo"
    elif "relación" in pregunta or "relacion" in pregunta or "parecidos" in pregunta or "similares" in pregunta:
        return "Descriptivo"
    elif "sí o no" in pregunta or "si o no" in pregunta:

    elif "comportamiento" in pregunta:
        return "Machine Learning: Clustering / Asociación"

        return "No se detectaron palabras clave claras"


preguntas_gimnasio = {
    "riesgo_baja": "¿Cuál es la probabilidad de que un socio se dé de baja este mes?",
    "perfil_socios": "¿Qué socios tienen hábitos de entrenamiento parecidos entre sí?",
    "renovacion": "¿Este socio va a renovar su membresía el próximo mes, sí o no?",
    "uso_instalaciones": "¿Qué patrones de comportamiento explican por qué algunos socios dejan de venir?",
}


def enfoque_para_preguntas(diccionario_preguntas):
    resultado = {}
    # Recorremos cada par (nombre_corto, pregunta) del diccionario de entrada

    return resultado
