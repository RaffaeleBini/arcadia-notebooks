### Soluzione dell'esercizio 1

```python
def calcular_psi(porcentaje_referencia, porcentaje_nuevo):
    return float(np.sum((porcentaje_nuevo - porcentaje_referencia) * np.log(porcentaje_nuevo / porcentaje_referencia)))

# Controllo automatico
psi_igual = calcular_psi(np.array([0.5, 0.5]), np.array([0.5, 0.5]))
assert abs(psi_igual - 0.0) < 1e-9, "Completa le righe contrassegnate con TODO"
psi_desplazado = calcular_psi(np.array([0.9, 0.1]), np.array([0.1, 0.9]))
assert psi_desplazado > 1.0, "Completa le righe contrassegnate con TODO"
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 2

```python
def interpretar_psi(psi):  # dato-es
    if psi < 0.10:
        return "sin drift"  # dato-es
    if psi < 0.25:
        return "drift moderado"  # dato-es
    return "drift severo"  # dato-es

# Controllo automatico
assert interpretar_psi(0.02) == "sin drift", "Completa le righe contrassegnate con TODO"  # dato-es
assert interpretar_psi(0.15) == "drift moderado", "Completa le righe contrassegnate con TODO"  # dato-es
assert interpretar_psi(0.30) == "drift severo", "Completa le righe contrassegnate con TODO"  # dato-es
assert interpretar_psi(0.10) == "drift moderado", "Completa le righe contrassegnate con TODO"  # dato-es
assert interpretar_psi(0.25) == "drift severo", "Completa le righe contrassegnate con TODO"  # dato-es
print("Esercizio 2 corretto ✔")
```

### Soluzione dell'esercizio 3

```python
def tipo_de_drift(cambia_distribucion_x, cambia_relacion_xy):
    if cambia_distribucion_x and cambia_relacion_xy:
        return "ambos"
    if cambia_distribucion_x:
        return "data drift"
    if cambia_relacion_xy:
        return "concept drift"
    return "ninguno"

# Controllo automatico
assert tipo_de_drift(True, False) == "data drift", "Completa le righe contrassegnate con TODO"
assert tipo_de_drift(False, True) == "concept drift", "Completa le righe contrassegnate con TODO"
assert tipo_de_drift(True, True) == "ambos", "Completa le righe contrassegnate con TODO"
assert tipo_de_drift(False, False) == "ninguno", "Completa le righe contrassegnate con TODO"
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
def variacion_relativa(valor_referencia, valor_nuevo):
    return round(valor_nuevo / valor_referencia, 2)

# Controllo automatico
assert variacion_relativa(0.9, 24.3) == 27.0, "Completa le righe contrassegnate con TODO"
assert variacion_relativa(10.0, 5.0) == 0.5, "Completa le righe contrassegnate con TODO"
assert variacion_relativa(2.0, 2.0) == 1.0, "Completa le righe contrassegnate con TODO"
print("Esercizio 4 corretto ✔")
```

### Soluzione dell'esercizio 5

```python
def debe_reentrenar(valores_psi, f1_actual, umbral_psi=0.25, umbral_f1=0.25):
    return any(psi > umbral_psi for psi in valores_psi) or f1_actual < umbral_f1

# Controllo automatico
assert debe_reentrenar([0.01, 0.02], 0.30) == False, "Completa le righe contrassegnate con TODO"
assert debe_reentrenar([0.01, 0.30], 0.30) == True, "Completa le righe contrassegnate con TODO"
assert debe_reentrenar([0.01, 0.02], 0.10) == True, "Completa le righe contrassegnate con TODO"
assert debe_reentrenar([], 0.05) == True, "Completa le righe contrassegnate con TODO"
print("Esercizio 5 corretto ✔")
```

### Soluzione della sfida finale

```python
def informe_drift(porcentajes_referencia, porcentajes_nuevos, f1_actual):
    psi_por_columna = [calcular_psi(ref, nuevo) for ref, nuevo in zip(porcentajes_referencia, porcentajes_nuevos)]
    interpretaciones = [interpretar_psi(psi) for psi in psi_por_columna]
    reentrenar = debe_reentrenar(psi_por_columna, f1_actual)
    return {"psi_por_columna": psi_por_columna, "interpretaciones": interpretaciones, "reentrenar": reentrenar}

# Controllo automatico
ref = [np.array([0.40, 0.30, 0.20, 0.10]), np.array([0.5, 0.5])]
nuevo = [np.array([0.10, 0.20, 0.30, 0.40]), np.array([0.5, 0.5])]
informe = informe_drift(ref, nuevo, f1_actual=0.30)
assert abs(informe["psi_por_columna"][0] - 0.91287) < 1e-4, "Completa le righe contrassegnate con TODO"
assert abs(informe["psi_por_columna"][1] - 0.0) < 1e-9, "Completa le righe contrassegnate con TODO"
assert informe["interpretaciones"] == ["drift severo", "sin drift"], "Completa le righe contrassegnate con TODO"  # dato-es
assert informe["reentrenar"] == True, "Completa le righe contrassegnate con TODO"

informe_sin_drift = informe_drift([np.array([0.5, 0.5])], [np.array([0.5, 0.5])], f1_actual=0.30)
assert informe_sin_drift["reentrenar"] == False, "Completa le righe contrassegnate con TODO"
print("Sfida finale corretta ✔")
```
