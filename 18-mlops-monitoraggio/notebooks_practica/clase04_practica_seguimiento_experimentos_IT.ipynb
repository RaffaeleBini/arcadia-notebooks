### Soluzione dell'esercizio 1

```python
def depende_del_umbral(nombre_metrica):
    return nombre_metrica in ("exactitud", "f1")

# Controllo automatico
assert depende_del_umbral("exactitud") is True, "Completa le righe contrassegnate con TODO"
assert depende_del_umbral("f1") is True, "Completa le righe contrassegnate con TODO"
assert depende_del_umbral("auc") is False, "Completa le righe contrassegnate con TODO"
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 2

```python
def f1_desde_precision_recall(precision, exhaustividad):
    if precision + exhaustividad == 0:
        return 0.0
    return 2 * precision * exhaustividad / (precision + exhaustividad)

# Controllo automatico
assert round(f1_desde_precision_recall(0.4, 0.25), 4) == 0.3077, "Completa le righe contrassegnate con TODO"
assert f1_desde_precision_recall(1.0, 1.0) == 1.0, "Completa le righe contrassegnate con TODO"
assert f1_desde_precision_recall(0.0, 0.0) == 0.0, "senza dividere per zero"
print("Esercizio 2 corretto ✔")
```

### Soluzione dell'esercizio 3

```python
def filtrar_por_auc(configuraciones, auc_minimo):
    return [c for c in configuraciones if c["auc"] >= auc_minimo]

# Controllo automatico
ejemplo = [{"id": 1, "auc": 0.72}, {"id": 2, "auc": 0.65}, {"id": 3, "auc": 0.70}]
resultado = filtrar_por_auc(ejemplo, 0.70)
assert [c["id"] for c in resultado] == [1, 3], "Completa le righe contrassegnate con TODO"
assert filtrar_por_auc(ejemplo, 0.9) == [], "se nessuno raggiunge il minimo, la lista deve restare vuota"
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
def desviacion(valores):
    media = sum(valores) / len(valores)
    return (sum((v - media) ** 2 for v in valores) / len(valores)) ** 0.5

# Controllo automatico
assert desviacion([5, 5, 5]) == 0.0, "Completa le righe contrassegnate con TODO"
assert round(desviacion([0.30, 0.32, 0.29]), 4) == 0.0125, "Completa le righe contrassegnate con TODO"
print("Esercizio 4 corretto ✔")
```

### Soluzione dell'esercizio 5

```python
def elegir_mejor(configuraciones, auc_minimo, f1_minimo):
    supera_auc = filtrar_por_auc(configuraciones, auc_minimo)
    supera_f1 = [c for c in supera_auc if c["f1"] >= f1_minimo]
    if not supera_f1:
        return None
    return min(supera_f1, key=lambda c: c["coste"])

# Controllo automatico
configs_prueba = [
    {"id": 1, "auc": 0.72, "f1": 0.0, "coste": 20},
    {"id": 3, "auc": 0.72, "f1": 0.32, "coste": 15},
    {"id": 4, "auc": 0.72, "f1": 0.31, "coste": 22},
    {"id": 6, "auc": 0.65, "f1": 0.30, "coste": 10},
]
elegido = elegir_mejor(configs_prueba, 0.70, 0.30)
assert elegido["id"] == 3, "Completa le righe contrassegnate con TODO"
assert elegir_mejor(configs_prueba, 0.90, 0.30) is None, "se nessuno supera l'AUC minimo, deve restituire None"
print("Esercizio 5 corretto ✔")
```

### Soluzione della sfida finale

```python
def configuracion_mas_estable(resultados_por_semilla, costes, umbral_estabilidad):
    estables = {
        config_id: f1s for config_id, f1s in resultados_por_semilla.items()
        if desviacion(f1s) <= umbral_estabilidad
    }
    if not estables:
        return None
    return max(estables, key=lambda config_id: sum(estables[config_id]) / len(estables[config_id]))

# Controllo automatico
resultados = {3: [0.3166, 0.3202, 0.2940], 4: [0.3138, 0.3258, 0.2902], 5: [0.3085, 0.3249, 0.2902]}
costes = {3: 15, 4: 22, 5: 16}
assert configuracion_mas_estable(resultados, costes, 0.02) == 3, "Completa le righe contrassegnate con TODO"
assert configuracion_mas_estable(resultados, costes, 0.001) is None, "con una soglia così rigida, nessuna configurazione dovrebbe essere abbastanza stabile"
print("Sfida finale corretta ✔")
```
