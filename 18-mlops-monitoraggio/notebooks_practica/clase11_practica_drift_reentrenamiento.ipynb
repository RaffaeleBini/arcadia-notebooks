### Solución del ejercicio 1

```python
def calcular_psi(porcentaje_referencia, porcentaje_nuevo):
    return float(np.sum((porcentaje_nuevo - porcentaje_referencia) * np.log(porcentaje_nuevo / porcentaje_referencia)))

# Comprobación automática
psi_igual = calcular_psi(np.array([0.5, 0.5]), np.array([0.5, 0.5]))
assert abs(psi_igual - 0.0) < 1e-9, "Completa las líneas marcadas con TODO"
psi_desplazado = calcular_psi(np.array([0.9, 0.1]), np.array([0.1, 0.9]))
assert psi_desplazado > 1.0, "Completa las líneas marcadas con TODO"
print("Ejercicio 1 correcto ✔")
```

### Solución del ejercicio 2

```python
def interpretar_psi(psi):  # dato-es
    if psi < 0.10:
        return "sin drift"  # dato-es
    if psi < 0.25:
        return "drift moderado"  # dato-es
    return "drift severo"  # dato-es

# Comprobación automática
assert interpretar_psi(0.02) == "sin drift", "Completa las líneas marcadas con TODO"  # dato-es
assert interpretar_psi(0.15) == "drift moderado", "Completa las líneas marcadas con TODO"  # dato-es
assert interpretar_psi(0.30) == "drift severo", "Completa las líneas marcadas con TODO"  # dato-es
assert interpretar_psi(0.10) == "drift moderado", "Completa las líneas marcadas con TODO"  # dato-es
assert interpretar_psi(0.25) == "drift severo", "Completa las líneas marcadas con TODO"  # dato-es
print("Ejercicio 2 correcto ✔")
```

### Solución del ejercicio 3

```python
def tipo_de_drift(cambia_distribucion_x, cambia_relacion_xy):
    if cambia_distribucion_x and cambia_relacion_xy:
        return "ambos"
    if cambia_distribucion_x:
        return "data drift"
    if cambia_relacion_xy:
        return "concept drift"
    return "ninguno"

# Comprobación automática
assert tipo_de_drift(True, False) == "data drift", "Completa las líneas marcadas con TODO"
assert tipo_de_drift(False, True) == "concept drift", "Completa las líneas marcadas con TODO"
assert tipo_de_drift(True, True) == "ambos", "Completa las líneas marcadas con TODO"
assert tipo_de_drift(False, False) == "ninguno", "Completa las líneas marcadas con TODO"
print("Ejercicio 3 correcto ✔")
```

### Solución del ejercicio 4

```python
def variacion_relativa(valor_referencia, valor_nuevo):
    return round(valor_nuevo / valor_referencia, 2)

# Comprobación automática
assert variacion_relativa(0.9, 24.3) == 27.0, "Completa las líneas marcadas con TODO"
assert variacion_relativa(10.0, 5.0) == 0.5, "Completa las líneas marcadas con TODO"
assert variacion_relativa(2.0, 2.0) == 1.0, "Completa las líneas marcadas con TODO"
print("Ejercicio 4 correcto ✔")
```

### Solución del ejercicio 5

```python
def debe_reentrenar(valores_psi, f1_actual, umbral_psi=0.25, umbral_f1=0.25):
    return any(psi > umbral_psi for psi in valores_psi) or f1_actual < umbral_f1

# Comprobación automática
assert debe_reentrenar([0.01, 0.02], 0.30) == False, "Completa las líneas marcadas con TODO"
assert debe_reentrenar([0.01, 0.30], 0.30) == True, "Completa las líneas marcadas con TODO"
assert debe_reentrenar([0.01, 0.02], 0.10) == True, "Completa las líneas marcadas con TODO"
assert debe_reentrenar([], 0.05) == True, "Completa las líneas marcadas con TODO"
print("Ejercicio 5 correcto ✔")
```

### Solución del reto final

```python
def informe_drift(porcentajes_referencia, porcentajes_nuevos, f1_actual):
    psi_por_columna = [calcular_psi(ref, nuevo) for ref, nuevo in zip(porcentajes_referencia, porcentajes_nuevos)]
    interpretaciones = [interpretar_psi(psi) for psi in psi_por_columna]
    reentrenar = debe_reentrenar(psi_por_columna, f1_actual)
    return {"psi_por_columna": psi_por_columna, "interpretaciones": interpretaciones, "reentrenar": reentrenar}

# Comprobación automática
ref = [np.array([0.40, 0.30, 0.20, 0.10]), np.array([0.5, 0.5])]
nuevo = [np.array([0.10, 0.20, 0.30, 0.40]), np.array([0.5, 0.5])]
informe = informe_drift(ref, nuevo, f1_actual=0.30)
assert abs(informe["psi_por_columna"][0] - 0.91287) < 1e-4, "Completa las líneas marcadas con TODO"
assert abs(informe["psi_por_columna"][1] - 0.0) < 1e-9, "Completa las líneas marcadas con TODO"
assert informe["interpretaciones"] == ["drift severo", "sin drift"], "Completa las líneas marcadas con TODO"  # dato-es
assert informe["reentrenar"] == True, "Completa las líneas marcadas con TODO"

informe_sin_drift = informe_drift([np.array([0.5, 0.5])], [np.array([0.5, 0.5])], f1_actual=0.30)
assert informe_sin_drift["reentrenar"] == False, "Completa las líneas marcadas con TODO"
print("Reto final correcto ✔")
```
