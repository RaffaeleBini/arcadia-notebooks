```python
def proporcion_nulos_aceptable(n_nulos, n_total, umbral_proporcion):
    return (n_nulos / n_total) <= umbral_proporcion

# Controllo automatico
assert proporcion_nulos_aceptable(3, 100, 0.05) is True, "Completa le righe contrassegnate con TODO"
assert proporcion_nulos_aceptable(10, 100, 0.05) is False, "Completa le righe contrassegnate con TODO"
print("Esercizio 2 corretto ✔")
```
