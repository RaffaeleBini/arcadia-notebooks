### Soluzione dell'esercizio 2

```python
def costo_por_1000_predicciones(latencia_ms, costo_hora_cpu):
    return round((costo_hora_cpu / 3600) * (latencia_ms / 1000) * 1000, 6)

# Controllo automatico
assert costo_por_1000_predicciones(4.0, 0.05) == 5.6e-05, "Completa le righe contrassegnate con TODO"
assert costo_por_1000_predicciones(0.0, 0.05) == 0.0, "Completa le righe contrassegnate con TODO"
assert costo_por_1000_predicciones(3600000.0, 1.0) == 1000.0, "Completa le righe contrassegnate con TODO"
print("Esercizio 2 corretto ✔")
```

### Soluzione dell'esercizio 3

```python
def costo_almacenamiento_mensual(bytes_totales, costo_por_gb_mes):
    return round((bytes_totales / 1024 ** 3) * costo_por_gb_mes, 8)

# Controllo automatico
assert costo_almacenamiento_mensual(1024 ** 3, 0.02) == 0.02, "Completa le righe contrassegnate con TODO"
assert costo_almacenamiento_mensual(0, 0.02) == 0.0, "Completa le righe contrassegnate con TODO"
assert costo_almacenamiento_mensual(157779, 0.02) == 2.94e-06, "Completa le righe contrassegnate con TODO"
print("Esercizio 3 corretto ✔")
```
