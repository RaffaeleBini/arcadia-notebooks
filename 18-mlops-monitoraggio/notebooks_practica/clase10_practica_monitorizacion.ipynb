### Solución del ejercicio 1

```python
def formatear_registro(endpoint, status_code, latencia_ms):
    registro = {"endpoint": endpoint, "status_code": status_code, "latencia_ms": latencia_ms}
    return json.dumps(registro, ensure_ascii=False)

# Comprobación automática
linea = formatear_registro("/predecir", 200, 12.3)
registro = json.loads(linea)
assert registro == {"endpoint": "/predecir", "status_code": 200, "latencia_ms": 12.3}, "Completa las líneas marcadas con TODO"
assert isinstance(linea, str), "Completa las líneas marcadas con TODO"
print("Ejercicio 1 correcto ✔")
```

### Solución del ejercicio 2

```python
def tasa_error(status_codes):
    if not status_codes:
        return 0.0
    return sum(1 for s in status_codes if s >= 500) / len(status_codes)

# Comprobación automática
assert tasa_error([200, 200, 200, 500]) == 0.25, "Completa las líneas marcadas con TODO"
assert tasa_error([200, 200, 200, 200]) == 0.0, "Completa las líneas marcadas con TODO"
assert tasa_error([500, 500]) == 1.0, "Completa las líneas marcadas con TODO"
print("Ejercicio 2 correcto ✔")
```

### Solución del ejercicio 3

```python
def percentil_95(latencias_ms):
    return float(np.percentile(latencias_ms, 95))

# Comprobación automática
valor = percentil_95([10, 10, 10, 10, 100])
assert abs(valor - float(np.percentile([10, 10, 10, 10, 100], 95))) < 1e-9, "Completa las líneas marcadas con TODO"
assert percentil_95([5, 5, 5, 5]) == 5.0, "Completa las líneas marcadas con TODO"
print("Ejercicio 3 correcto ✔")
```

### Solución del ejercicio 4

```python
def detectar_incidentes_simple(panel, umbral_latencia_ms, umbral_tasa_error):
    horas = [
        fila["hora"] for fila in panel
        if fila["latencia_media_ms"] > umbral_latencia_ms or fila["tasa_error"] > umbral_tasa_error
    ]
    return sorted(horas)

# Comprobación automática
panel_prueba = [
    {"hora": 0, "latencia_media_ms": 12.0, "tasa_error": 0.01},
    {"hora": 1, "latencia_media_ms": 11.5, "tasa_error": 0.02},
    {"hora": 2, "latencia_media_ms": 150.0, "tasa_error": 0.03},
    {"hora": 3, "latencia_media_ms": 12.5, "tasa_error": 0.40},
]
assert detectar_incidentes_simple(panel_prueba, 50.0, 0.10) == [2, 3], "Completa las líneas marcadas con TODO"
assert detectar_incidentes_simple(panel_prueba, 500.0, 0.90) == [], "Completa las líneas marcadas con TODO"
print("Ejercicio 4 correcto ✔")
```

### Solución del ejercicio 5

```python
def contar_falsos_positivos(horas_marcadas, horas_incidente_real):
    return len([h for h in horas_marcadas if h not in horas_incidente_real])

# Comprobación automática
assert contar_falsos_positivos([2, 3, 7, 14], {2, 14}) == 2, "Completa las líneas marcadas con TODO"
assert contar_falsos_positivos([2, 14], {2, 14}) == 0, "Completa las líneas marcadas con TODO"
assert contar_falsos_positivos([], {2, 14}) == 0, "Completa las líneas marcadas con TODO"
print("Ejercicio 5 correcto ✔")
```

### Solución del reto final

```python
def evaluar_umbral(panel, umbral_latencia_ms, umbral_tasa_error, horas_incidente_real):
    detectados = detectar_incidentes_simple(panel, umbral_latencia_ms, umbral_tasa_error)
    falsos_positivos = contar_falsos_positivos(detectados, horas_incidente_real)
    incidentes_perdidos = len([h for h in horas_incidente_real if h not in detectados])
    return {"detectados": detectados, "falsos_positivos": falsos_positivos, "incidentes_perdidos": incidentes_perdidos}

# Comprobación automática
panel_reto = [
    {"hora": 0, "latencia_media_ms": 12.0, "tasa_error": 0.01},
    {"hora": 1, "latencia_media_ms": 13.0, "tasa_error": 0.02},
    {"hora": 2, "latencia_media_ms": 150.0, "tasa_error": 0.03},
    {"hora": 3, "latencia_media_ms": 12.5, "tasa_error": 0.40},
]
horas_reales = {2, 3}
informe = evaluar_umbral(panel_reto, 50.0, 0.10, horas_reales)
assert informe == {"detectados": [2, 3], "falsos_positivos": 0, "incidentes_perdidos": 0}, "Completa las líneas marcadas con TODO"

informe_agresivo = evaluar_umbral(panel_reto, 10.0, 0.015, horas_reales)
assert informe_agresivo["detectados"] == [0, 1, 2, 3], "Completa las líneas marcadas con TODO"
assert informe_agresivo["falsos_positivos"] == 2, "Completa las líneas marcadas con TODO"
assert informe_agresivo["incidentes_perdidos"] == 0, "Completa las líneas marcadas con TODO"

informe_alto = evaluar_umbral(panel_reto, 500.0, 0.90, horas_reales)
assert informe_alto["detectados"] == [], "Completa las líneas marcadas con TODO"
assert informe_alto["falsos_positivos"] == 0, "Completa las líneas marcadas con TODO"
assert informe_alto["incidentes_perdidos"] == 2, "Completa las líneas marcadas con TODO"
print("Reto final correcto ✔")
```
