## ¿Por qué nos importa esto?

Ya sabes guardar datos en listas y tuplas. Pero en ciencia de datos aparecen todo el tiempo dos problemas que las listas resuelven mal:

1. **Eliminar duplicados** — ¿cuántos géneros musicales distintos hay en un dataset de 10.000 canciones?
2. **Buscar por identificador, no por posición** — ¿cuál es el precio del producto `"Laptop"`? En una lista tendrías que recordar en qué índice está.

Para el primer problema usamos **sets**. Para el segundo, **diccionarios**.

**Regla práctica:**
- Lista → cuando el orden y la posición importan.
- Set → cuando importa la unicidad o comparar dos grupos entre sí.
- Diccionario → cuando importa buscar por nombre/identificador.


## 1. Sets: colecciones de valores únicos

Un **set** es una colección de objetos únicos, sin orden y sin duplicados. Se define con llaves `{}`, igual que un diccionario, pero solo con valores (sin claves). Python descarta los duplicados automáticamente.

```python
# Crear un set con géneros repetidos
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
print(set1)
# 'rock' aparece una sola vez, aunque lo escribimos tres veces
```

Piénsalo como una caja de canicas de colores: no importa cuántas canicas rojas metas, dentro de la caja solo existe "el color rojo" una vez.

Si ya tienes una lista con duplicados, la conviertes a set con `set()`:

```python
# Convertir una lista en un set (elimina duplicados y mezcla tipos distintos)
album_list = ["Michael Jackson", "Thriller", 1982, "00:42:19",
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
print(album_set)
```

```python
# Otro ejemplo: lista de géneros con repeticiones -> set de géneros únicos
music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul",
                     "progressive rock", "soft rock", "R&B", "disco"])
print(music_genres)
```

**Importante:** como un set no tiene orden, no puedes acceder a sus elementos por posición (`set1[0]` da error). Solo puedes preguntar si algo está dentro, añadir, quitar y comparar con otros sets.

### 1.1 Agregar, quitar y comprobar elementos

```python
A = set(["Thriller", "Back in Black", "AC/DC"])
print(A)

# Agregar un elemento con add()
A.add("NSYNC")
print(A)
```

```python
# Agregar el mismo elemento otra vez no hace nada: no hay duplicados
A.add("NSYNC")
print(A)
```

```python
# Quitar un elemento con remove()
A.remove("NSYNC")
print(A)

# Comprobar si un elemento está en el set con 'in'
print("AC/DC" in A)   # True
```

`remove()` lanza un error si el elemento no existe. `discard()` hace lo mismo pero sin romper el programa si el elemento no está — útil cuando no estás seguro de si algo ya fue eliminado.

### 1.2 Operaciones de conjuntos (diagrama de Venn)

Imagina dos discotecas con distintas playlists. Queremos saber qué canciones tienen en común, cuáles son exclusivas de cada una, y cuál sería la playlist combinada.

```python
album_set1 = set(["Thriller", "AC/DC", "Back in Black"])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
print(album_set1)
print(album_set2)
```

Ambos sets comparten `"AC/DC"` y `"Back in Black"`. Esa zona donde se superponen los dos círculos del Venn es la **intersección**: se calcula con `&` o con `.intersection()`.

```python
# Intersección: lo que está en ambos sets
interseccion = album_set1 & album_set2
print(interseccion)

# Forma equivalente con método
print(album_set1.intersection(album_set2))
```

La **diferencia** da lo que está en un set pero no en el otro — el pedazo de círculo que no se superpone. El orden importa.

```python
# Lo que solo está en album_set1
print(album_set1.difference(album_set2))   # {'Thriller'}

# Lo que solo está en album_set2
print(album_set2.difference(album_set1))   # {'The Dark Side of the Moon'}
```

La **unión** junta todo, coloreando los dos círculos completos, sin repetir lo que aparece en ambos.

```python
# Unión: todo junto, sin duplicados
print(album_set1.union(album_set2))
```

Por último, `issubset()` (subconjunto) e `issuperset()` (superconjunto) preguntan si un set está completamente "contenido" dentro de otro.

```python
# ¿Es album_set2 subconjunto de album_set1? (todos sus elementos están en album_set1)
print(album_set2.issubset(album_set1))   # False

# ¿Es album_set1 superconjunto de {"Back in Black", "AC/DC"}?
print(album_set1.issuperset({"Back in Black", "AC/DC"}))   # True
```

**Resumen de sets:** `union()` (todo junto), `intersection()` (lo común), `difference()` (lo exclusivo de uno) y `issubset()`/`issuperset()` (relación de contención).

## 2. Diccionarios: pares clave-valor

Un diccionario guarda pares **clave-valor**: en vez de acceder por posición numérica (como en una lista), accedes por un nombre que tú eliges. El ejemplo clásico es un número de identificación: ese número único es la **clave**, y los datos asociados son el **valor**.

```python
# Crear un diccionario
release_year_dict = {
    "Thriller": "1982",
    "Back in Black": "1980",
    "The Dark Side of the Moon": "1973",
    "The Bodyguard": "1992",
    "Bat Out of Hell": "1977",
    "Their Greatest Hits (1971-1975)": "1976",
    "Saturday Night Fever": "1977",
    "Rumours": "1977",
}
print(release_year_dict)
```

Cada clave va separada de su valor con `:`, las entradas se separan con comas, y todo va entre `{}`. Un diccionario vacío se escribe `{}`.

¿Cuándo diccionario y cuándo lista? Compara:

```python
# Con lista: ¿qué es cada cosa? Hay que recordar el orden
persona_lista = ["Ana", 34, "Madrid"]

# Con diccionario: autoexplicativo
persona_dict = {"nombre": "Ana", "edad": 34, "ciudad": "Madrid"}
print(persona_dict["edad"])  # 34, sin ambigüedad
```

Las **claves** pueden ser strings, números, o tuplas (porque son inmutables). No pueden ser listas ni otro diccionario. Los **valores**, en cambio, pueden ser de cualquier tipo: números, strings, listas, tuplas, incluso otros diccionarios.

```python
Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4),
        "key5": 5, (0, 1): 6}

print(Dict["key1"])   # 1
print(Dict[(0, 1)])   # 6 -- una tupla también sirve como clave
```

### 2.1 keys(), values() e items(): recorrer un diccionario

```python
# Todas las claves
print(release_year_dict.keys())

# Todos los valores
print(release_year_dict.values())
```

```python
# Si quieres una lista normal en vez de esa vista especial, envuélvelo en list()
claves = list(release_year_dict.keys())
valores = list(release_year_dict.values())
print(claves)
print(valores)
```

```python
# items() da tuplas (clave, valor)
pares = list(release_year_dict.items())
print(pares)
```

### 2.2 Agregar, modificar y eliminar entradas

```python
# Acceder a un valor por su clave
print(release_year_dict["Thriller"])
print(release_year_dict["The Bodyguard"])
```

```python
# Agregar una entrada nueva
release_year_dict["Graduation"] = "2007"

# Modificar una entrada existente (misma sintaxis)
release_year_dict["Thriller"] = "1982 (reedición)"

print(release_year_dict)
```

```python
# Fusionar varias entradas de golpe con update()
release_year_dict.update({"Random Access Memories": "2013"})
print(release_year_dict)
```

```python
# Eliminar una entrada con del
del release_year_dict["Thriller"]
del release_year_dict["Graduation"]
print(release_year_dict)
```

```python
# Comprobar si una clave existe con 'in'
print("The Bodyguard" in release_year_dict)   # True
print("Thriller" in release_year_dict)        # False, ya la borramos
```

Dos métodos útiles para tener en el radar:
- `clear()` vacía el diccionario entero sin borrar la variable.
- `copy()` da una copia independiente: modificar la copia no toca al original.

## Cierre: resumen de la clase

- Un **set** es una colección de valores únicos y sin orden; se escribe entre `{}` y elimina duplicados automáticamente.
- Con un set: `add()` para agregar, `remove()`/`discard()` para quitar, `in` para comprobar pertenencia. No hay acceso por posición.
- Operaciones de conjuntos: `union()`, `intersection()`, `difference()`, `issubset()`, `issuperset()` — como comparar círculos en un diagrama de Venn.
- Un **diccionario** guarda pares clave-valor: `diccionario["clave"]`. Se escribe entre `{}` con `clave: valor` separados por comas.
- Claves inmutables (strings, números, tuplas); valores de cualquier tipo.
- `keys()`, `values()`, `items()` recorren el diccionario.
- Asignar a clave nueva la crea, a clave existente la sobrescribe. `del diccionario[clave]` elimina, `clear()` vacía todo, `in` comprueba existencia.
- **Regla práctica:** lista si el orden importa, set si la unicidad/comparación importa, diccionario si buscas por nombre o identificador.
