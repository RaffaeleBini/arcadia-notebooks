## 1. Clase vs. instancia: el plano y la casa

Una **clase** es como el plano de una casa: define que va a tener paredes, puertas y ventanas, pero el plano en sí no es habitable. Cuando construís una casa real a partir de ese plano, esa casa es una **instancia**. Podés construir muchas casas distintas a partir del mismo plano, cada una con sus propios valores (color de puerta, por ejemplo).

En Python, la clase es el plano y el objeto (instancia) es la casa real.

```python
# La clase es la plantilla: define qué atributos y métodos va a tener
# cualquier círculo, sin crear ninguno todavía.
class Circle(object):
    pass

# Cada línea siguiente crea una instancia (un objeto) distinto,
# aunque ambas vengan del mismo plano (la clase Circle).
RedCircle = Circle()
BlueCircle = Circle()

print(type(RedCircle))
print(type(BlueCircle))
```

`Circle` es un tipo, igual que `str` o `list`. `RedCircle` es un valor de ese tipo, igual que `"hola"` es un valor de tipo `str`.

### `dir()`: inspeccionar qué trae un objeto por dentro

## 2. El constructor `__init__` y `self`

Una clase vacía no sirve de mucho: no tiene atributos. Para darle atributos a cada instancia en el momento de crearla, usamos un método especial llamado **constructor**, que en Python siempre se llama `__init__`. Python lo ejecuta automáticamente cada vez que creás un objeto de esa clase.

El constructor recibe `self` como primer parámetro **siempre**. `self` es una referencia a la instancia concreta que se está creando. Cuando escribís `self.radius = radius`, le decís a Python "guardá este valor como atributo *de este objeto en particular*". Nunca lo escribís al llamar al método —Python lo pasa solo—, pero sí tenés que declararlo como primer parámetro en la definición.

```python
import matplotlib

matplotlib.use('Agg')  # backend sin ventana, para poder ejecutar el notebook sin entorno gráfico

import matplotlib.pyplot as plt

%matplotlib inline



class Circle(object):



    # Constructor: se ejecuta al crear cada instancia

    def __init__(self, radius=3, color='blue'):

        self.radius = radius

        self.color = color
```

```python
RedCircle = Circle(10, 'red')



print('radius:', RedCircle.radius)

print('color:', RedCircle.color)
```

Cuando ejecutás `Circle(10, 'red')`, Python crea un objeto nuevo y llama a `Circle.__init__(nuevo_objeto, 10, 'red')` por debajo. Vos nunca pasás `self` a mano; Python lo completa con la instancia recién creada.

Sin `self`, Python no tendría forma de distinguir "el radio de este círculo" de "el radio de aquel otro". Es lo que permite que cada instancia mantenga sus propios valores, aunque provengan de la misma clase.

## 3. Atributos y métodos

Los **atributos** (`self.radius`, `self.color`) son las variables que cuelgan de cada instancia y describen su estado. Se acceden y modifican con la notación de punto:

```python
RedCircle = Circle(10, 'red')



print(RedCircle.radius)   # 10

print(RedCircle.color)    # red



RedCircle.radius = 1      # modificamos el atributo directamente

print(RedCircle.radius)   # 1
```

Los **métodos** son funciones definidas dentro de la clase que operan sobre esos atributos. Se diferencian de una función común en que siempre reciben `self` como primer parámetro, porque necesitan saber sobre qué instancia están actuando.

```python
class Circle(object):



    def __init__(self, radius=3, color='blue'):

        self.radius = radius

        self.color = color



    # Método: modifica un atributo del objeto

    def add_radius(self, r):

        self.radius = self.radius + r

        return self.radius



    # Método: dibuja el objeto usando sus propios atributos

    def drawCircle(self):

        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))

        plt.axis('scaled')

        plt.show()
```

```python
RedCircle = Circle(10, 'red')



print('Radio inicial:', RedCircle.radius)

RedCircle.add_radius(2)

print('Después de +2:', RedCircle.radius)

RedCircle.add_radius(5)

print('Después de +5:', RedCircle.radius)



RedCircle.drawCircle()
```

Cada llamada a `add_radius` opera sobre el estado interno de `RedCircle` específicamente. Si tuvieras otro círculo, `BlueCircle`, llamar a `BlueCircle.add_radius(2)` no afectaría en nada a `RedCircle`: cada instancia guarda su propia copia de los atributos.

El foco de `drawCircle` no es matplotlib en sí, sino que **usa los atributos guardados** (`self.radius`, `self.color`) sin que se los tengas que volver a pasar: el objeto ya sabe cómo dibujarse.

La misma estructura se repite con cualquier otra clase, como `Rectangle`:

```python
class Rectangle(object):



    def __init__(self, width=2, height=3, color='r'):

        self.height = height

        self.width = width

        self.color = color



    def drawRectangle(self):

        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height, fc=self.color))

        plt.axis('scaled')

        plt.show()



SkinnyBlueRectangle = Rectangle(2, 3, 'blue')

FatYellowRectangle = Rectangle(20, 5, 'yellow')



print(SkinnyBlueRectangle.width)    # 2

print(FatYellowRectangle.color)     # yellow



SkinnyBlueRectangle.drawRectangle()

FatYellowRectangle.drawRectangle()
```

Dos instancias de la misma clase, con atributos totalmente distintos, comportándose de acuerdo al mismo plano.

## 4. Valores por defecto y atributos de clase

Los parámetros del constructor pueden tener **valores por defecto**, igual que en cualquier función: si no especificás ese argumento al crear la instancia, Python usa el valor por defecto.

```python
BlueCircle = Circle(radius=100)   # color queda en 'blue', el valor por defecto



print(BlueCircle.color)    # blue

print(BlueCircle.radius)   # 100
```

Los valores por defecto no tienen que vivir solo en el constructor. También podés definir un **atributo de clase** —un valor compartido por todas las instancias a menos que se sobreescriba— escribiéndolo directamente dentro de la clase, fuera del constructor.

```python
class Vehicle:

    color = "white"   # atributo de clase: todas las instancias parten de "white"



    def __init__(self, max_speed, mileage):

        self.max_speed = max_speed

        self.mileage = mileage



auto1 = Vehicle(200, 50000)

auto2 = Vehicle(180, 75000)



print(auto1.color, auto2.color)   # ambos parten de 'white'



auto1.color = "rojo"              # sobreescribimos solo en esta instancia

print(auto1.color, auto2.color)
```

`color` no depende de lo que le pases al constructor: todo vehículo nuevo arranca siendo blanco, salvo que después cambies el atributo sobre una instancia puntual. La diferencia con `max_speed` es que ese sí es un atributo de **instancia** — cada vehículo lo recibe como argumento y lo guarda por separado —, mientras que `color` es un valor por defecto compartido a nivel de **clase**.

Ahora armemos una clase `Vehicle` completa, combinando atributos de instancia, un atributo de clase y varios métodos — el sistema de inventario de un concesionario:

```python
class Vehicle:

    color = "white"



    def __init__(self, max_speed, mileage):

        self.max_speed = max_speed

        self.mileage = mileage

        self.seating_capacity = None



    def assign_seating_capacity(self, seating_capacity):

        self.seating_capacity = seating_capacity



    def display_properties(self):

        print("Propiedades del vehículo:")

        print("Color:", self.color)

        print("Velocidad máxima:", self.max_speed)

        print("Kilometraje:", self.mileage)

        print("Capacidad de asientos:", self.seating_capacity)



vehicle1 = Vehicle(200, 50000)

vehicle1.assign_seating_capacity(5)

vehicle1.display_properties()



print()



vehicle2 = Vehicle(180, 75000)

vehicle2.assign_seating_capacity(4)

vehicle2.display_properties()
```

`vehicle1` y `vehicle2` son dos instancias completamente independientes: cada una guarda su propio `max_speed`, `mileage` y `seating_capacity`, aunque compartan el atributo de clase `color`. `assign_seating_capacity` es un método que no recibe el dato en el constructor, sino que lo asigna después, cuando lo necesitás — algo perfectamente válido en POO: no todos los atributos tienen que fijarse en el momento de crear el objeto.

## 5. Caso combinado: `TextAnalyzer`

Todo lo anterior cobra sentido cuando lo aplicás a algo que ya sabés hacer sin clases, y ves cuánto se ordena. Supongamos que analizas reseñas de clientes y querés: (1) normalizar el texto (minúsculas, sin puntuación), (2) calcular la frecuencia de cada palabra, y (3) consultar la frecuencia de una palabra específica.

Con clases, el texto formateado se calcula **una sola vez**, en el constructor, y queda guardado como atributo — los demás métodos simplemente lo reutilizan.

```python
class TextAnalyzer(object):



    def __init__(self, text):

        # Quitar puntuación: reutilizamos lo que ya sabemos de strings

        formattedText = text.replace('.', '').replace('!', '').replace('?', '').replace(',', '')



        # Pasar a minúsculas

        formattedText = formattedText.lower()



        # Guardar el resultado como atributo de la instancia

        self.fmtText = formattedText



    def freqAll(self):

        # Separar el texto en palabras

        wordList = self.fmtText.split(' ')



        # Construir un diccionario de frecuencias, como en el capítulo de diccionarios

        freqMap = {}

        for word in set(wordList):   # set() elimina duplicados de la lista

            freqMap[word] = wordList.count(word)



        return freqMap



    def freqOf(self, word):

        # Reutilizar freqAll() en vez de recalcular todo de nuevo

        freqDict = self.freqAll()



        if word in freqDict:

            return freqDict[word]

        else:

            return 0
```

Tres decisiones de diseño son el corazón de este ejemplo:

- El constructor hace el trabajo de limpieza **una sola vez** y lo guarda en `self.fmtText`. Ningún otro método vuelve a recibir el texto crudo como parámetro.
- `freqAll()` no recibe nada más que `self`, porque todo lo que necesita (`self.fmtText`) ya está en el objeto.
- `freqOf(word)` no repite la lógica de conteo: llama a `self.freqAll()` y reutiliza su resultado. Un método puede llamar a otro método del mismo objeto usando `self.`.

```python
givenstring = "Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet."



analyzed = TextAnalyzer(givenstring)



print("Texto formateado:", analyzed.fmtText)



freqMap = analyzed.freqAll()

print(freqMap)



word = "lorem"

frequency = analyzed.freqOf(word)

print("La palabra", word, "aparece", frequency, "veces.")
```

Una sola llamada, `TextAnalyzer(givenstring)`, deja el objeto listo con el texto ya limpio guardado adentro. A partir de ahí, cada método que llamás sobre `analyzed` trabaja con ese estado interno sin que vos tengas que volver a pasarle el texto. Esto es exactamente lo que ganás con POO frente a funciones sueltas: el objeto **recuerda su propio estado** entre una llamada y la siguiente.

## Resumen: lo que te llevas de esta clase

- Una **clase** es el plano que define qué atributos y métodos va a tener un tipo de objeto; una **instancia** (u objeto) es una realización concreta de ese plano, con sus propios valores.
- El **constructor** `__init__` se ejecuta automáticamente al crear una instancia y es donde normalmente asignás los atributos iniciales.
- `self` es la referencia a la instancia concreta sobre la que se está trabajando. Todo método de una clase lo recibe como primer parámetro, aunque nunca lo pases explícitamente al llamarlo.
- Los **atributos** guardan el estado del objeto (`self.radius`, `self.color`); los **métodos** son funciones que operan sobre ese estado y siempre llevan `self`.
- Accedés y modificás atributos, y llamás métodos, con la notación de punto: `objeto.atributo`, `objeto.metodo(argumentos)`.
- Podés dar **valores por defecto** a los parámetros del constructor y también definir **atributos de clase**, compartidos por todas las instancias salvo que los sobreescribas.
- Un método puede llamar a otro método del mismo objeto usando `self.otro_metodo()`, evitando repetir lógica — así lo hace `freqOf` al reutilizar `freqAll`.
- `dir(objeto)` te muestra qué métodos y atributos tiene disponibles, incluidos los que Python agrega por defecto.
- La POO no reemplaza strings, listas ni diccionarios: los organiza. `TextAnalyzer` es una clase por fuera, pero por dentro sigue usando `.replace()`, `.lower()`, `.split()` y un diccionario de frecuencias.
