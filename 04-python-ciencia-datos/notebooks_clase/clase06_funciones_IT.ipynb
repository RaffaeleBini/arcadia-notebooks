## 1. Definire una funzione: `def`, `return` e docstring

Pensa a una funzione come a una ricetta di cucina: ha un nome, una lista di ingredienti (i **parametri**) e un risultato finale (ciò che **`return`** restituisce).

In Python definisci una funzione con `def`, il nome, le parentesi con i parametri, e i due punti. Il corpo va indentato:

- **`def add(a):`** — il nome è `add`, riceve un parametro `a`.
- Il **docstring** (`"""..."""` subito dopo i due punti) documenta cosa fa la funzione. Python lo usa per `help(nome_funzione)`.
- **`return b`** — termina la funzione e "restituisce" il valore di `b` a chi l'ha chiamata.

```python
def add(a):
    """
    Somma 1 ad 'a' e restituisce il risultato.
    """
    b = a + 1
    print(a, "se ci sommi uno, dà", b)
    return b


add(1)
add(2)
```

Grazie al docstring, possiamo chiedere aiuto a Python su questa funzione:

```python
help(add)
```

Un dettaglio importante: non appena Python esegue `return`, la funzione **termina lì**, anche se restano righe di codice dopo.

```python
def Mult(a, b):
    c = a * b
    return c
    print('Questo non viene mai stampato')  # il return ha già interrotto l'esecuzione


resultado = Mult(12, 2)
print(resultado)  # 24
```

Se una funzione non ha nessun `return`, Python le restituisce automaticamente `None`. Nota la differenza tra **stampare** (mostrare qualcosa a schermo) e **restituire** (fornire un valore utilizzabile al codice che ha chiamato la funzione):

```python
def MJ():
    print('Michael Jackson')


def MJ1():
    print('Michael Jackson')
    return None


print(MJ())    # stampa il testo, e poi stampa None (ciò che restituisce MJ)
print(MJ1())   # esattamente lo stesso, return None è il comportamento predefinito
```

Una funzione può anche combinare tutto quello che già conosci sulle strutture di controllo: un `if` che decide cosa restituire, o un `for` che scorre una lista che le passi come parametro.

```python
def type_of_album(artist, album, year_released):
    print(artist, album, year_released)
    if year_released > 1980:
        return "Moderno"
    else:
        return "Antiguo"


x = type_of_album("Michael Jackson", "Thriller", 1980)
print(x)  # Antiguo
```

```python
def PrintList(the_list):
    for element in the_list:
        print(element)


PrintList(['1', 1, 'the man', "abc"])
```

Un esempio più completo, che mette insieme `def`, un ciclo, un dizionario e `.split()`: contare quante volte compare ogni parola in un testo.

```python
def freq(string):
    # Passo 1: separiamo il testo in una lista di parole
    words = string.split()

    # Passo 2: dizionario vuoto per accumulare i conteggi
    Dict = {}

    # Passo 3: per ogni parola, contiamo quante volte compare in tutta la lista
    for key in words:
        Dict[key] = words.count(key)

    print("La frequenza delle parole è:", Dict)


freq("Mary had a little lamb Little lamb, little lamb Mary had a little lamb")
```

Questa funzione mette insieme praticamente tutto ciò che hai visto nei capitoli precedenti — stringhe, liste, dizionari, cicli — dentro un unico blocco riutilizzabile. Questo è proprio il punto di una funzione: il luogo dove combini pezzi separati per risolvere un problema concreto, e gli dai un nome per non doverci più ripensare.

## 2. Parametri e valori predefiniti

Alcune ricette hanno ingredienti opzionali: se non specifichi nulla, si usa una quantità standard. In Python questo è un **valore predefinito**, e si definisce direttamente nella firma della funzione:

```python
def isGoodRating(rating=4):
    if rating < 7:
        print("questo disco è brutto, il suo voto è", rating)
    else:
        print("questo disco è bello, il suo voto è", rating)


isGoodRating()      # usa rating=4 -> "questo disco è brutto, il suo voto è 4"
isGoodRating(10)     # usa rating=10 -> "questo disco è bello, il suo voto è 10"
```

Questo è utile quando la maggior parte delle volte chiamerai la funzione con lo stesso valore tipico, ma vuoi lasciare aperta la possibilità di cambiarlo in casi specifici.

⚠️ **Regola**: i parametri con valore predefinito vanno sempre **dopo** quelli che non ce l'hanno. `def f(a=1, b):` è un errore in Python.

## 3. Scope: variabili locali e globali

Questa è la parte dove più persone si confondono, quindi andiamo con calma. Pensa a ogni funzione come a una stanza con la porta chiusa: ciò che crei al suo interno (una **variabile locale**) resta dentro e scompare quando la funzione termina.

```python
def square(a):
    b = 1              # b è una variabile locale: esiste solo qui dentro
    c = a * a + b
    print(a, "al quadrato più 1 fa", c)
    return c


x = 3
y = square(x)
print(y)     # 10 -> questo funziona, "y" è una variabile normale fuori dalla funzione
```

Se proviamo ad accedere a `b` fuori dalla funzione, otteniamo un errore, perché `b` non è mai esistita fuori da `square`:

```python
try:
    print(b)
except NameError as e:
    print("Errore:", e)
```

Invece, una **variabile globale** si definisce fuori da qualsiasi funzione, ed è visibile da qualunque parte del programma, incluso l'interno delle funzioni (per *leggerla*):

```python
myFavouriteBand = "AC/DC"   # variabile globale


def getBandRating(bandname):
    if bandname == myFavouriteBand:   # la funzione può LEGGERE la variabile globale
        return 10.0
    else:
        return 0.0


print("Il voto di AC/DC è:", getBandRating("AC/DC"))
print("La mia band preferita è:", myFavouriteBand)
```

Ora la parte più insidiosa: se dentro una funzione crei una variabile con lo **stesso nome** di una globale, Python **non tocca la globale**. Crea una nuova variabile locale che esiste solo dentro quella funzione, e "copre" (fa *shadowing* di) la globale unicamente dentro quel blocco.

```python
myFavouriteBand = "AC/DC"          # globale


def getBandRating(bandname):
    myFavouriteBand = "Deep Purple"  # questa è una nuova variabile LOCALE, non quella sopra
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0


print("Il voto dei Deep Purple è:", getBandRating("Deep Purple"))  # 10.0
print("La mia band preferita resta:", myFavouriteBand)             # AC/DC
```

E se hai davvero bisogno che una funzione **modifichi** una variabile globale? Per questo esiste la parola chiave **`global`**: dice a Python "questa variabile non è locale, è quella di fuori (o voglio crearla come globale)".

```python
artist = "Michael Jackson"


def printer(artist):
    global internal_var
    internal_var = "Whitney Houston"
    print(artist, "è un artista")


printer(artist)
printer(internal_var)   # ora internal_var esiste davvero fuori dalla funzione
```

Senza quel `global`, `internal_var` sarebbe stata una variabile locale che scompare al termine della funzione, e la seconda riga avrebbe dato un `NameError`.

**Consiglio sincero**: usa `global` con moderazione. È quasi sempre meglio che una funzione riceva ciò di cui ha bisogno come parametro e restituisca il suo risultato con `return`, invece di leggere e scrivere variabili globali di nascosto.

## 4. `*args` e `**kwargs`: quando non sai quanti argomenti arriveranno

Cosa fai se vuoi una funzione che accetti "tutti gli argomenti che le passi", senza sapere in anticipo quanti saranno?

**`*args`** impacchetta tutti gli argomenti sciolti (posizionali) in una **tupla**:

```python
def printAll(*args):
    print("Numero di argomenti:", len(args))
    for argument in args:
        print(argument)


printAll('Horsefeather', 'Adonis', 'Bone')
print()
printAll('Sidecar', 'Long Island', 'Mudslide', 'Carriage')  # funziona uguale con 4 argomenti
```

Il nome `args` è solo una convenzione (potresti chiamarlo `*qualsiasi_cosa`); ciò che dice a Python "impacchetta qui" è l'asterisco `*`.

**`**kwargs`** fa lo stesso ma per argomenti `chiave=valore`, impacchettati in un **dizionario**:

```python
def printDictionary(**args):
    for key in args:
        print(key + " : " + args[key])


printDictionary(Country='Canada', Province='Ontario', City='Toronto')
```

Quando useresti questo? Quando scrivi una funzione "generica" che deve accettare un numero variabile di valori senza anticipare tutti i casi possibili — ad esempio, una funzione che riceve un numero qualsiasi di colonne di un dataset, o configurazioni opzionali con nome. Le vedrai continuamente nella documentazione di librerie come `pandas`.

## 5. Passare una lista a una funzione: attenzione alla mutabilità

Quando passi un numero o una stringa a una funzione, la funzione riceve una **copia** del valore. Ma le liste (e i dizionari) funzionano diversamente: quando passi una lista, passi un **riferimento** allo stesso oggetto in memoria. Se la funzione la modifica, il cambiamento si vede anche fuori.

```python
def addItems(lista):
    lista.append("Three")
    lista.append("Four")


myList = ["One", "Two"]
addItems(myList)

print(myList)   # ['One', 'Two', 'Three', 'Four']
```

`addItems` non ha nessun `return`, eppure `myList` è cambiata fuori dalla funzione. Questo accade perché `lista` (il parametro) e `myList` (la variabile esterna) puntano **allo stesso oggetto**: non ci sono "due liste", c'è un'unica lista con due nomi che vi puntano.

Questo è molto diverso da `square(a)`, dove `a` era un numero: numeri, stringhe e tuple sono **immutabili**, non si possono modificare "sul posto".

Perché conta nella Data Science? Perché lavorerai costantemente con liste, dizionari e, più avanti, DataFrame di `pandas`, che si comportano allo stesso modo. Se una funzione modifica "per errore" un oggetto pensando che sia una copia, puoi corrompere i dati originali. La soluzione, quando ti serve che la funzione non tocchi l'originale, è fare una copia esplicita con `.copy()` prima di modificarla.

```python
def addItemsSeguro(lista):
    copia = lista.copy()   # lavoriamo su una copia, non sull'originale
    copia.append("Three")
    copia.append("Four")
    return copia


myList2 = ["One", "Two"]
nuevaLista = addItemsSeguro(myList2)

print("Originale (intatta):", myList2)
print("Nuova lista:", nuevaLista)
```

## Riepilogo: cosa porti a casa da questa lezione

- Una funzione è un blocco di codice riutilizzabile: si definisce una volta con `def nome(parametri):` e si chiama tutte le volte che serve.
- `return` termina la funzione e restituisce un valore; senza `return`, la funzione restituisce `None`. Stampare con `print()` non è lo stesso che restituire con `return`.
- Puoi documentare una funzione con un docstring e consultarlo con `help(nome_funzione)`.
- Una variabile creata **dentro** una funzione è **locale**; una creata **fuori** da qualsiasi funzione è **globale** ed è visibile in tutto il programma, incluse le funzioni (per leggerla).
- Se dentro una funzione crei una variabile con lo stesso nome di una globale, si crea una nuova variabile locale che non tocca la globale — a meno che tu non usi `global nome_variabile`.
- `*args` impacchetta argomenti posizionali variabili in una tupla; `**kwargs` impacchetta argomenti con nome variabili in un dizionario.
- I parametri con valore predefinito (`def f(x=4):`) vanno sempre dopo i parametri senza valore predefinito.
- Passare una lista o un dizionario a una funzione passa un **riferimento**, non una copia: se la funzione la modifica, il cambiamento si vede anche fuori. Usa `.copy()` quando devi evitarlo.
