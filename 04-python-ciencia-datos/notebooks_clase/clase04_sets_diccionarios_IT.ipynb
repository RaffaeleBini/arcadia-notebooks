## Perché ci interessa?

Sai già come salvare dati in liste e tuple. Ma nella data science compaiono continuamente due problemi che le liste risolvono male:

1. **Eliminare i duplicati** — quanti generi musicali distinti ci sono in un dataset di 10.000 canzoni?
2. **Cercare per identificativo, non per posizione** — qual è il prezzo del prodotto `"Laptop"`? In una lista dovresti ricordare a quale indice si trova.

Per il primo problema usiamo i **set**. Per il secondo, i **dizionari**.

**Regola pratica:**
- Lista → quando l'ordine e la posizione contano.
- Set → quando conta l'unicità o il confronto tra due gruppi.
- Dizionario → quando conta la ricerca per nome/identificativo.


## 1. Set: collezioni di valori unici

Un **set** è una collezione di oggetti unici, senza ordine e senza duplicati. Si definisce con le parentesi graffe `{}`, come un dizionario, ma con soli valori (senza chiavi). Python scarta automaticamente i duplicati.

```python
# Creare un set con generi ripetuti
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
print(set1)
# 'rock' compare una sola volta, anche se lo abbiamo scritto tre volte
```

Pensalo come una scatola di biglie colorate: non importa quante biglie rosse ci metti dentro, nella scatola esiste "il colore rosso" una sola volta.

Se hai già una lista con duplicati, la converti in set con `set()`:

```python
# Convertire una lista in un set (elimina i duplicati e mescola tipi diversi)
album_list = ["Michael Jackson", "Thriller", 1982, "00:42:19",
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
print(album_set)
```

```python
# Un altro esempio: lista di generi con ripetizioni -> set di generi unici
music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul",
                     "progressive rock", "soft rock", "R&B", "disco"])
print(music_genres)
```

**Importante:** siccome un set non ha ordine, non puoi accedere ai suoi elementi per posizione (`set1[0]` dà errore). Puoi solo chiedere se qualcosa è presente, aggiungere, rimuovere e confrontare con altri set.

### 1.1 Aggiungere, rimuovere e verificare elementi

```python
A = set(["Thriller", "Back in Black", "AC/DC"])
print(A)

# Aggiungere un elemento con add()
A.add("NSYNC")
print(A)
```

```python
# Aggiungere lo stesso elemento un'altra volta non fa nulla: non ci sono duplicati
A.add("NSYNC")
print(A)
```

```python
# Rimuovere un elemento con remove()
A.remove("NSYNC")
print(A)

# Verificare se un elemento è nel set con 'in'
print("AC/DC" in A)   # True
```

`remove()` lancia un errore se l'elemento non esiste. `discard()` fa lo stesso ma senza interrompere il programma se l'elemento non c'è — utile quando non sei sicuro se qualcosa è già stato eliminato.

### 1.2 Operazioni sugli insiemi (diagramma di Venn)

Immagina due discoteche con playlist diverse. Vogliamo sapere quali canzoni hanno in comune, quali sono esclusive di ciascuna, e quale sarebbe la playlist combinata.

```python
album_set1 = set(["Thriller", "AC/DC", "Back in Black"])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])
print(album_set1)
print(album_set2)
```

Entrambi i set condividono `"AC/DC"` e `"Back in Black"`. Quella zona dove si sovrappongono i due cerchi del Venn è l'**intersezione**: si calcola con `&` o con `.intersection()`.

```python
# Intersezione: ciò che è presente in entrambi i set
interseccion = album_set1 & album_set2
print(interseccion)

# Forma equivalente con metodo
print(album_set1.intersection(album_set2))
```

La **differenza** dà ciò che è in un set ma non nell'altro — la parte di cerchio che non si sovrappone. L'ordine conta.

```python
# Ciò che è presente solo in album_set1
print(album_set1.difference(album_set2))   # {'Thriller'}

# Ciò che è presente solo in album_set2
print(album_set2.difference(album_set1))   # {'The Dark Side of the Moon'}
```

L'**unione** mette insieme tutto, colorando i due cerchi per intero, senza ripetere ciò che compare in entrambi.

```python
# Unione: tutto insieme, senza duplicati
print(album_set1.union(album_set2))
```

Infine, `issubset()` (sottoinsieme) e `issuperset()` (soprainsieme) chiedono se un set è completamente "contenuto" in un altro.

```python
# album_set2 è un sottoinsieme di album_set1? (tutti i suoi elementi sono in album_set1)
print(album_set2.issubset(album_set1))   # False

# album_set1 è un soprainsieme di {"Back in Black", "AC/DC"}?
print(album_set1.issuperset({"Back in Black", "AC/DC"}))   # True
```

**Riepilogo dei set:** `union()` (tutto insieme), `intersection()` (ciò in comune), `difference()` (ciò esclusivo di uno) e `issubset()`/`issuperset()` (relazione di contenimento).

## 2. Dizionari: coppie chiave-valore

Un dizionario salva coppie **chiave-valore**: invece di accedere per posizione numerica (come in una lista), accedi tramite un nome che scegli tu. L'esempio classico è un numero di identificazione: quel numero univoco è la **chiave**, e i dati associati sono il **valore**.

```python
# Creare un dizionario
release_year_dict = {
    "Thriller": "1982",
    "Back in Black": "1980",
    "The Dark Side of the Moon": "1973",
    "The Bodyguard": "1992",
    "Bat Out of Hell": "1977",
    "Their Greatest Hits (1971-1975)": "1976",
    "Saturday Night Fever": "1977",
    "Rumours": "1977",
}
print(release_year_dict)
```

Ogni chiave è separata dal suo valore con `:`, le voci sono separate da virgole, e tutto va tra `{}`. Un dizionario vuoto si scrive `{}`.

Quando usare un dizionario e quando una lista? Confronta:

```python
# Con lista: cos'è ogni cosa? Bisogna ricordare l'ordine
persona_lista = ["Ana", 34, "Madrid"]

# Con dizionario: autoesplicativo
persona_dict = {"nombre": "Ana", "edad": 34, "ciudad": "Madrid"}
print(persona_dict["edad"])  # 34, senza ambiguità
```

Le **chiavi** possono essere stringhe, numeri, o tuple (perché sono immutabili). Non possono essere liste né un altro dizionario. I **valori**, invece, possono essere di qualsiasi tipo: numeri, stringhe, liste, tuple, persino altri dizionari.

```python
Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4),
        "key5": 5, (0, 1): 6}

print(Dict["key1"])   # 1
print(Dict[(0, 1)])   # 6 -- anche una tupla può servire come chiave
```

### 2.1 keys(), values() e items(): scorrere un dizionario

```python
# Tutte le chiavi
print(release_year_dict.keys())

# Tutti i valori
print(release_year_dict.values())
```

```python
# Se vuoi una lista normale invece di questa vista speciale, avvolgila in list()
claves = list(release_year_dict.keys())
valores = list(release_year_dict.values())
print(claves)
print(valores)
```

```python
# items() dà tuple (chiave, valore)
pares = list(release_year_dict.items())
print(pares)
```

### 2.2 Aggiungere, modificare ed eliminare voci

```python
# Accedere a un valore tramite la sua chiave
print(release_year_dict["Thriller"])
print(release_year_dict["The Bodyguard"])
```

```python
# Aggiungere una nuova voce
release_year_dict["Graduation"] = "2007"

# Modificare una voce esistente (stessa sintassi)
release_year_dict["Thriller"] = "1982 (riedizione)"

print(release_year_dict)
```

```python
# Fondere più voci in un colpo solo con update()
release_year_dict.update({"Random Access Memories": "2013"})
print(release_year_dict)
```

```python
# Eliminare una voce con del
del release_year_dict["Thriller"]
del release_year_dict["Graduation"]
print(release_year_dict)
```

```python
# Verificare se una chiave esiste con 'in'
print("The Bodyguard" in release_year_dict)   # True
print("Thriller" in release_year_dict)        # False, l'abbiamo già cancellata
```

Due metodi utili da tenere a mente:
- `clear()` svuota l'intero dizionario senza cancellare la variabile.
- `copy()` restituisce una copia indipendente: modificare la copia non tocca l'originale.

## Chiusura: riepilogo della lezione

- Un **set** è una collezione di valori unici e senza ordine; si scrive tra `{}` ed elimina automaticamente i duplicati.
- Con un set: `add()` per aggiungere, `remove()`/`discard()` per rimuovere, `in` per verificare l'appartenenza. Non c'è accesso per posizione.
- Operazioni sugli insiemi: `union()`, `intersection()`, `difference()`, `issubset()`, `issuperset()` — come confrontare cerchi in un diagramma di Venn.
- Un **dizionario** salva coppie chiave-valore: `dizionario["chiave"]`. Si scrive tra `{}` con `chiave: valore` separati da virgole.
- Chiavi immutabili (stringhe, numeri, tuple); valori di qualsiasi tipo.
- `keys()`, `values()`, `items()` scorrono il dizionario.
- Assegnare a una chiave nuova la crea, a una chiave esistente la sovrascrive. `del dizionario[chiave]` elimina, `clear()` svuota tutto, `in` verifica l'esistenza.
- **Regola pratica:** lista se l'ordine conta, set se conta l'unicità/il confronto, dizionario se cerchi per nome o identificativo.
