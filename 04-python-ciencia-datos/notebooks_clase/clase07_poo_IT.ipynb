## 1. Classe vs. istanza: il progetto e la casa

Una **classe** è come il progetto di una casa: definisce che avrà pareti, porte e finestre, ma il progetto in sé non è abitabile. Quando costruisci una casa reale a partire da quel progetto, quella casa è un'**istanza**. Puoi costruire molte case diverse a partire dallo stesso progetto, ciascuna con i propri valori (colore della porta, ad esempio).

In Python, la classe è il progetto e l'oggetto (istanza) è la casa reale.

```python
# La classe è il modello: definisce quali attributi e metodi avrà
# qualsiasi cerchio, senza crearne ancora nessuno.
class Circle(object):
    pass

# Ogni riga seguente crea un'istanza (un oggetto) diversa,
# anche se entrambe provengono dallo stesso progetto (la classe Circle).
RedCircle = Circle()
BlueCircle = Circle()

print(type(RedCircle))
print(type(BlueCircle))
```

`Circle` è un tipo, proprio come `str` o `list`. `RedCircle` è un valore di quel tipo, proprio come `"ciao"` è un valore di tipo `str`.

### `dir()`: ispezionare cosa contiene un oggetto al suo interno

`dir()` elenca i metodi e gli attributi disponibili di un oggetto. Vedrai un mix di metodi propri della tua classe e metodi predefiniti che Python aggiunge a tutti gli oggetti (quelli che iniziano e finiscono con doppio underscore, come `__init__`). Non serve memorizzarli: `dir()` è il tuo strumento di ispezione per quando non ricordi cosa può fare un oggetto.

```python
# Vediamo cosa contiene al suo interno l'oggetto RedCircle

dir(RedCircle)
```

## 2. Il costruttore `__init__` e `self`

Una classe vuota non serve a molto: non ha attributi. Per dare attributi a ogni istanza nel momento in cui viene creata, usiamo un metodo speciale chiamato **costruttore**, che in Python si chiama sempre `__init__`. Python lo esegue automaticamente ogni volta che crei un oggetto di quella classe.

Il costruttore riceve `self` come primo parametro **sempre**. `self` è un riferimento all'istanza concreta che si sta creando. Quando scrivi `self.radius = radius`, dici a Python "salva questo valore come attributo *di questo oggetto in particolare*". Non lo scrivi mai quando chiami il metodo — Python lo passa da solo — ma devi dichiararlo come primo parametro nella definizione.

```python
import matplotlib

matplotlib.use('Agg')  # backend senza finestra, per poter eseguire il notebook senza ambiente grafico

import matplotlib.pyplot as plt

%matplotlib inline



class Circle(object):



    # Costruttore: viene eseguito alla creazione di ogni istanza

    def __init__(self, radius=3, color='blue'):

        self.radius = radius

        self.color = color
```

```python
RedCircle = Circle(10, 'red')



print('radius:', RedCircle.radius)

print('color:', RedCircle.color)
```

Quando esegui `Circle(10, 'red')`, Python crea un nuovo oggetto e chiama `Circle.__init__(nuovo_oggetto, 10, 'red')` sotto il cofano. Tu non passi mai `self` a mano; Python lo completa con l'istanza appena creata.

Senza `self`, Python non avrebbe modo di distinguere "il raggio di questo cerchio" da "il raggio di quell'altro". È ciò che permette a ogni istanza di mantenere i propri valori, anche se provengono dalla stessa classe.

## 3. Attributi e metodi

Gli **attributi** (`self.radius`, `self.color`) sono le variabili appese a ogni istanza e che ne descrivono lo stato. Si accedono e modificano con la notazione a punto:

```python
RedCircle = Circle(10, 'red')



print(RedCircle.radius)   # 10

print(RedCircle.color)    # red



RedCircle.radius = 1      # modifichiamo l'attributo direttamente

print(RedCircle.radius)   # 1
```

I **metodi** sono funzioni definite dentro la classe che operano su quegli attributi. Si differenziano da una funzione comune perché ricevono sempre `self` come primo parametro, perché hanno bisogno di sapere su quale istanza stanno agendo.

```python
class Circle(object):



    def __init__(self, radius=3, color='blue'):

        self.radius = radius

        self.color = color



    # Metodo: modifica un attributo dell'oggetto

    def add_radius(self, r):

        self.radius = self.radius + r

        return self.radius



    # Metodo: disegna l'oggetto usando i propri attributi

    def drawCircle(self):

        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))

        plt.axis('scaled')

        plt.show()
```

```python
RedCircle = Circle(10, 'red')



print('Raggio iniziale:', RedCircle.radius)

RedCircle.add_radius(2)

print('Dopo +2:', RedCircle.radius)

RedCircle.add_radius(5)

print('Dopo +5:', RedCircle.radius)



RedCircle.drawCircle()
```

Ogni chiamata a `add_radius` opera sullo stato interno di `RedCircle` in modo specifico. Se avessi un altro cerchio, `BlueCircle`, chiamare `BlueCircle.add_radius(2)` non influenzerebbe in alcun modo `RedCircle`: ogni istanza conserva la propria copia degli attributi.

Il punto centrale di `drawCircle` non è matplotlib in sé, ma il fatto che **usa gli attributi salvati** (`self.radius`, `self.color`) senza che tu debba passarglieli di nuovo: l'oggetto sa già come disegnarsi.

La stessa struttura si ripete con qualsiasi altra classe, come `Rectangle`:

```python
class Rectangle(object):



    def __init__(self, width=2, height=3, color='r'):

        self.height = height

        self.width = width

        self.color = color



    def drawRectangle(self):

        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height, fc=self.color))

        plt.axis('scaled')

        plt.show()



SkinnyBlueRectangle = Rectangle(2, 3, 'blue')

FatYellowRectangle = Rectangle(20, 5, 'yellow')



print(SkinnyBlueRectangle.width)    # 2

print(FatYellowRectangle.color)     # yellow



SkinnyBlueRectangle.drawRectangle()

FatYellowRectangle.drawRectangle()
```

Due istanze della stessa classe, con attributi completamente diversi, che si comportano secondo lo stesso progetto.

## 4. Valori predefiniti e attributi di classe

I parametri del costruttore possono avere **valori predefiniti**, proprio come in qualsiasi funzione: se non specifichi quell'argomento nel creare l'istanza, Python usa il valore predefinito.

```python
BlueCircle = Circle(radius=100)   # color rimane 'blue', il valore predefinito



print(BlueCircle.color)    # blue

print(BlueCircle.radius)   # 100
```

I valori predefiniti non devono vivere solo nel costruttore. Puoi anche definire un **attributo di classe** — un valore condiviso da tutte le istanze a meno che non venga sovrascritto — scrivendolo direttamente dentro la classe, fuori dal costruttore.

```python
class Vehicle:

    color = "white"   # attributo di classe: tutte le istanze partono da "white"



    def __init__(self, max_speed, mileage):

        self.max_speed = max_speed

        self.mileage = mileage



auto1 = Vehicle(200, 50000)

auto2 = Vehicle(180, 75000)



print(auto1.color, auto2.color)   # entrambi partono da 'white'



auto1.color = "rosso"             # sovrascriviamo solo su questa istanza

print(auto1.color, auto2.color)
```

`color` non dipende da ciò che passi al costruttore: ogni nuovo veicolo parte bianco, a meno che poi non cambi l'attributo su un'istanza specifica. La differenza con `max_speed` è che quello sì è un attributo di **istanza** — ogni veicolo lo riceve come argomento e lo salva separatamente —, mentre `color` è un valore predefinito condiviso a livello di **classe**.

Ora costruiamo una classe `Vehicle` completa, combinando attributi di istanza, un attributo di classe e vari metodi — il sistema di inventario di un concessionario:

```python
class Vehicle:

    color = "white"



    def __init__(self, max_speed, mileage):

        self.max_speed = max_speed

        self.mileage = mileage

        self.seating_capacity = None



    def assign_seating_capacity(self, seating_capacity):

        self.seating_capacity = seating_capacity



    def display_properties(self):

        print("Proprietà del veicolo:")

        print("Colore:", self.color)

        print("Velocità massima:", self.max_speed)

        print("Chilometraggio:", self.mileage)

        print("Capacità posti:", self.seating_capacity)



vehicle1 = Vehicle(200, 50000)

vehicle1.assign_seating_capacity(5)

vehicle1.display_properties()



print()



vehicle2 = Vehicle(180, 75000)

vehicle2.assign_seating_capacity(4)

vehicle2.display_properties()
```

`vehicle1` e `vehicle2` sono due istanze completamente indipendenti: ciascuna conserva il proprio `max_speed`, `mileage` e `seating_capacity`, anche se condividono l'attributo di classe `color`. `assign_seating_capacity` è un metodo che non riceve il dato nel costruttore, ma lo assegna dopo, quando serve — un approccio perfettamente valido nella OOP: non tutti gli attributi devono essere fissati nel momento in cui si crea l'oggetto.

## 5. Caso combinato: `TextAnalyzer`

Tutto quanto sopra acquista senso quando lo applichi a qualcosa che già sai fare senza classi, e vedi quanto si ordina. Supponiamo che tu stia analizzando le recensioni dei clienti e voglia: (1) normalizzare il testo (minuscolo, senza punteggiatura), (2) calcolare la frequenza di ogni parola, e (3) consultare la frequenza di una parola specifica.

Con le classi, il testo formattato viene calcolato **una sola volta**, nel costruttore, e resta salvato come attributo — gli altri metodi lo riutilizzano semplicemente.

```python
class TextAnalyzer(object):



    def __init__(self, text):

        # Rimuovere la punteggiatura: riutilizziamo ciò che sappiamo già sulle stringhe

        formattedText = text.replace('.', '').replace('!', '').replace('?', '').replace(',', '')



        # Convertire in minuscolo

        formattedText = formattedText.lower()



        # Salvare il risultato come attributo dell'istanza

        self.fmtText = formattedText



    def freqAll(self):

        # Separare il testo in parole

        wordList = self.fmtText.split(' ')



        # Costruire un dizionario di frequenze, come nel capitolo sui dizionari

        freqMap = {}

        for word in set(wordList):   # set() elimina i duplicati dalla lista

            freqMap[word] = wordList.count(word)



        return freqMap



    def freqOf(self, word):

        # Riutilizzare freqAll() invece di ricalcolare tutto da capo

        freqDict = self.freqAll()



        if word in freqDict:

            return freqDict[word]

        else:

            return 0
```

Tre scelte di design sono il cuore di questo esempio:

- Il costruttore fa il lavoro di pulizia **una sola volta** e lo salva in `self.fmtText`. Nessun altro metodo riceve di nuovo il testo grezzo come parametro.
- `freqAll()` non riceve nient'altro che `self`, perché tutto ciò di cui ha bisogno (`self.fmtText`) è già nell'oggetto.
- `freqOf(word)` non ripete la logica di conteggio: chiama `self.freqAll()` e riutilizza il suo risultato. Un metodo può chiamare un altro metodo dello stesso oggetto usando `self.`.

```python
givenstring = "Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet."



analyzed = TextAnalyzer(givenstring)



print("Testo formattato:", analyzed.fmtText)



freqMap = analyzed.freqAll()

print(freqMap)



word = "lorem"

frequency = analyzed.freqOf(word)

print("La parola", word, "compare", frequency, "volte.")
```

Una sola chiamata, `TextAnalyzer(givenstring)`, lascia l'oggetto pronto con il testo già ripulito salvato al suo interno. Da quel momento, ogni metodo che chiami su `analyzed` lavora con quello stato interno senza che tu debba passargli di nuovo il testo. Questo è esattamente ciò che guadagni con la OOP rispetto a funzioni separate: l'oggetto **ricorda il proprio stato** tra una chiamata e la successiva.

## Riepilogo: cosa porti a casa da questa lezione

- Una **classe** è il progetto che definisce quali attributi e metodi avrà un tipo di oggetto; un'**istanza** (o oggetto) è una realizzazione concreta di quel progetto, con i propri valori.
- Il **costruttore** `__init__` viene eseguito automaticamente alla creazione di un'istanza ed è dove normalmente assegni gli attributi iniziali.
- `self` è il riferimento all'istanza concreta su cui si sta lavorando. Ogni metodo di una classe lo riceve come primo parametro, anche se non lo passi mai esplicitamente quando lo chiami.
- Gli **attributi** conservano lo stato dell'oggetto (`self.radius`, `self.color`); i **metodi** sono funzioni che operano su quello stato e portano sempre `self`.
- Accedi e modifichi gli attributi, e chiami i metodi, con la notazione a punto: `oggetto.attributo`, `oggetto.metodo(argomenti)`.
- Puoi dare **valori predefiniti** ai parametri del costruttore e anche definire **attributi di classe**, condivisi da tutte le istanze a meno che tu non li sovrascriva.
- Un metodo può chiamare un altro metodo dello stesso oggetto usando `self.altro_metodo()`, evitando di ripetere la logica — è quello che fa `freqOf` riutilizzando `freqAll`.
- `dir(oggetto)` ti mostra quali metodi e attributi ha disponibili, inclusi quelli che Python aggiunge di default.
- La OOP non sostituisce stringhe, liste o dizionari: li organizza. `TextAnalyzer` è una classe all'esterno, ma internamente continua a usare `.replace()`, `.lower()`, `.split()` e un dizionario di frequenze.
