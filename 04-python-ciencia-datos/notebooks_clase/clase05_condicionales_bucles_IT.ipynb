## Perché ci interessa?

Finora il tuo codice viene eseguito sempre allo stesso modo, dall'alto verso il basso. Ma lavorare con dati reali significa **prendere decisioni** ("questo valore è valido?") e **ripetere processi** ("fai questo per ogni riga"). Condizionali e cicli sono i due strumenti che ti permettono di fare entrambe le cose, e quasi sempre si usano insieme: scorri una collezione con un ciclo e, al suo interno, decidi con un `if` cosa fare con ogni elemento.

## 1. Operatori di confronto

| Operatore | Significato |
|---|---|
| `==` | uguale a |
| `!=` | diverso da |
| `>` | maggiore di |
| `<` | minore di |
| `>=` | maggiore o uguale a |
| `<=` | minore o uguale a |

⚠️ **Attenzione**: `=` assegna un valore, `==` confronta due valori. È l'errore più comune quando si inizia.

```python
edad = 25
print(edad == 25)   # l'età vale 25?
print(edad != 30)   # l'età è diversa da 30?
print(edad >= 20)   # l'età è maggiore o uguale a 20?
```

Questi operatori funzionano anche con le **stringhe**. Python confronta in base al codice ASCII di ogni carattere (le maiuscole valgono meno delle minuscole), e il confronto è sensibile a maiuscole/minuscole.

```python
print("ACDC" == "Michael Jackson")   # False, sono stringhe diverse
print("ACDC" != "Michael Jackson")   # True

print('B' > 'A')     # True -> 66 > 65 in ASCII
print('BA' > 'AB')   # True -> confronta prima la prima lettera ('B' > 'A')
```

## 2. `if` / `else`: prendere decisioni

Pensa a un `if` come a una porta con serratura: se la condizione è `True`, la porta si apre e il blocco indentato viene eseguito; se è `False`, viene ignorato completamente, senza generare alcun errore.

```python
edad = 19

if edad > 18:
    print("puoi entrare al concerto")

print("il programma continua")   # questo viene eseguito sempre, porta aperta o chiusa
```

`else` copre il caso in cui la condizione dell'`if` è risultata falsa. Non ha una condizione propria: è il "in qualsiasi altro caso".

```python
edad = 17

if edad > 18:
    print("puoi entrare al concerto rock")
else:
    print("meglio andare a vedere il film")
```

### `elif`: più di due strade

Quando ci sono più di due possibilità, concateni condizioni con `elif` ("else if"). Python le valuta **in ordine** ed esegue il primo blocco vero; ignora il resto, anche se fosse ugualmente vero.

```python
edad = 18

if edad > 18:
    print("vai al concerto A (ACDC)")
elif edad == 18:
    print("vai al concerto B (Pink Floyd)")   # questa è quella eseguita
else:
    print("vai al concerto C (Meat Loaf)")

print("continui la tua giornata")
```

Un uso molto tipico di `elif` concatenati: classificare per fasce, come la colazione di una mensa scolastica in base all'età dello studente.

```python
edad = 10

if edad <= 9:
    print("riceve pappa d'avena")
elif edad >= 10 and edad <= 14:
    print("riceve un panino")
elif edad >= 15 and edad <= 17:
    print("riceve un hamburger")
```

### Condizionali annidati

Puoi mettere un `if` dentro un altro `if` (ad esempio, per validare un dettaglio dentro un ramo principale, come in un bancomat). È perfettamente valido, ma se ti ritrovi ad annidare 4 o 5 livelli, probabilmente puoi semplificare combinando condizioni con operatori logici — il prossimo argomento.

```python
opcion = "Retirar efectivo"

if opcion == "Retirar efectivo":
    monto = 50
    if monto % 10 == 0:
        print("Importo erogato:", monto)
    else:
        print("Inserisci un multiplo di 10.")
else:
    print("Grazie per aver usato il bancomat.")
```

## 3. Operatori logici: `and`, `or`, `not`

- **`and`** — tutte le condizioni devono essere soddisfatte (come una serratura che richiede tessera *e* impronta digitale).
- **`or`** — basta che una condizione sia soddisfatta (come scegliere un film tra amici: se a qualcuno piace il genere, si sceglie).
- **`not`** — inverte il risultato di una condizione.

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
tiene_tarjeta = True
tiene_huella = True

if tiene_tarjeta and tiene_huella:
    print("accesso consentito")

le_gusta_comedia = True
le_gusta_accion = False
le_gusta_drama = False

if le_gusta_comedia or le_gusta_accion or le_gusta_drama:
    print("si sceglie un film")

no_molestar = True

if not no_molestar:
    print("invia notifica")
else:
    print("nessuna notifica inviata, modalità non disturbare attiva")
```

Un uso molto frequente nella data science è verificare se un valore rientra **dentro** o **fuori** di un intervallo, combinando due confronti con `and` oppure `or`.

```python
anio_album = 1985

if anio_album > 1979 and anio_album < 1990:
    print("l'album è degli anni 80")

anio_album = 1995

if anio_album < 1980 or anio_album > 1989:
    print("l'album non è degli anni 80")
else:
    print("l'album è degli anni 80")
```

## 4. Il ciclo `for`: ripetere su una sequenza

Usa `for` quando **sai in anticipo** su quale sequenza andrai a iterare (una lista, una stringa, un `range()`) e vuoi elaborare ciascuno dei suoi elementi.

```python
colores = ["rosso", "arancione", "giallo", "verde", "blu", "indaco", "viola"]

for color in colores:
    print(color)
```

### `range()`: generare sequenze di numeri

- `range(n)` genera numeri da `0` a `n - 1`.
- `range(inicio, fin)` genera numeri da `inicio` a `fin - 1` (il limite superiore non è mai incluso).

```python
for numero in range(11):
    print(numero)   # stampa da 0 a 10

print("---")

for numero in range(1, 11):
    print(numero)   # stampa da 1 a 10
```

`range(len(lista))` serve per scorrere una lista **per indice** invece che per valore, utile quando ti serve quell'indice per qualcos'altro — ad esempio, per modificare la lista mentre la scorri.

```python
cuadrados = ['rosso', 'giallo', 'verde', 'viola', 'blu']

for i in range(len(cuadrados)):
    print("prima:", cuadrados[i])
    cuadrados[i] = 'bianco'
    print("dopo:", cuadrados[i])
```

### Esempio reale: tabelline

Un uso classico di `for` + `range()`: generare tabelline complete senza scrivere ogni riga a mano.

```python
print("Tabellina del 6:")
for i in range(10):
    print("6 *", i, "=", 6 * i)

print("\nTabellina del 7:")
for i in range(10):
    print("7 *", i, "=", 7 * i)
```

### `enumerate()`: l'indice e il valore insieme

Se ti serve l'indice solo per sapere "in che posizione sono" ma vuoi anche il valore, `enumerate()` ti restituisce entrambi a ogni giro ed è più leggibile ("pythonico") che tenere tu stesso il conto con un contatore.

```python
frutas = ["mela", "banana", "arancia"]

for indice, fruta in enumerate(frutas):
    print(f"in posizione {indice} c'è {fruta}")
```

## 5. Il ciclo `while`: ripetere finché una condizione è vera

Usa `while` quando **non sai in anticipo** quanti giri ti serviranno: ripeti un'operazione finché una condizione resta vera.

```python
contador = 1
while contador <= 10:
    print(contador)
    contador += 1   # senza questo, la condizione non cambia mai e il ciclo non finisce mai
```

Un esempio più realistico: scorrere una lista di anni e fermarti appena ne trovi uno in particolare. Qui `while` è lo strumento corretto perché non sai in anticipo in quale posizione si trova 1973 — la condizione di arresto dipende dal contenuto dei dati, non da un conteggio fisso.

```python
anios = [1982, 1980, 1973, 2000]

i = 0
anio_actual = anios[0]

while anio_actual != 1973:
    print(anio_actual)
    i += 1
    anio_actual = anios[i]

print("ci sono voluti", i, "giri per trovare 1973")
```

### Esempio reale: filtrare animali per lunghezza del nome

Un altro caso molto tipico: scorrere una collezione e costruire una nuova lista solo con gli elementi che soddisfano una condizione.

```python
animales = ["leon", "jirafa", "gorila", "loros", "cocodrilo", "ciervo", "cisne"]
animales_siete_letras = []

i = 0
while i < len(animales):
    if len(animales[i]) == 7:
        animales_siete_letras.append(animales[i])
    i += 1

print(animales_siete_letras)
```

## 6. `for` vs. `while`: quale scelgo?

- **Usa `for`** quando conosci in anticipo la sequenza su cui andrai a iterare (una lista completa, un intervallo, ogni carattere di una stringa). Il numero di giri è determinato dalla dimensione della sequenza.
- **Usa `while`** quando il numero di ripetizioni dipende da una condizione conosciuta solo in fase di esecuzione ("finché non trovi il valore", "finché il rating non scende sotto 6").

Regola pratica: se puoi rispondere "ripeterò questo esattamente N volte, o una volta per ogni elemento di questa collezione", usa `for`. Se la tua risposta è "ripeterò questo finché non succede qualcosa", usa `while` — e assicurati sempre che "quel qualcosa" abbia modo di verificarsi, o il ciclo non finirà mai.

```python
# Lo stesso compito risolto con for e con while, per confrontare

# Sommare i primi 5 numeri con for (sappiamo esattamente quanti giri servono)
suma_for = 0
for n in range(1, 6):
    suma_for += n
print("somma con for:", suma_for)

# La stessa somma con while (funziona, ma qui for è più diretto)
suma_while = 0
n = 1
while n <= 5:
    suma_while += n
    n += 1
print("somma con while:", suma_while)
```

## Riepilogo della lezione

- `==`, `!=`, `>`, `<`, `>=`, `<=` restituiscono `True`/`False` e sono la base di qualsiasi decisione. Non confondere `==` (confrontare) con `=` (assegnare).
- `if` esegue un blocco solo se la sua condizione è vera; `else` copre tutti gli altri casi; `elif` concatena altre condizioni e Python esegue la prima che si verifica.
- L'indentazione definisce quale codice appartiene a un blocco `if`/`elif`/`else` — non è un dettaglio estetico, è sintassi.
- `and` richiede che tutte le condizioni siano vere, `or` richiede che almeno una lo sia, `not` inverte il risultato.
- `for` scorre una sequenza conosciuta; `range(n)` genera numeri da `0` a `n-1`; `enumerate()` ti dà indice e valore insieme.
- `while` ripete finché una condizione è vera; usalo quando non sai in anticipo quante ripetizioni serviranno — e non dimenticare mai di aggiornare la variabile della condizione, o avrai un ciclo infinito.
- Condizionali e cicli quasi sempre si combinano: scorri una collezione con un ciclo e, al suo interno, decidi con un `if` cosa fare con ogni elemento — lo schema che userai per pulire e filtrare dati nei prossimi capitoli.
