## 3. Errori comuni: `NameError` e `SyntaxError`

Poiché Python esegue riga per riga, se c'è un errore in una riga, tutto quello precedente è già stato eseguito normalmente e l'esecuzione si ferma proprio lì.

- **`NameError`**: hai usato un nome (funzione o variabile) che Python non riconosce.
- **`SyntaxError`**: hai infranto una regola "grammaticale" del linguaggio (ad esempio, una virgoletta non chiusa).

Vediamo un `NameError` in azione — scriviamo male `print` come `frint`. **Questa cella fallirà di proposito**, è l'obiettivo dell'esempio:

```python
# Questo genera un NameError di proposito (frint non esiste)
frint("Hello, Python!")
```

Ora un `SyntaxError` — ci siamo dimenticati di chiudere la virgoletta. **Anche questa cella fallirà di proposito**:

```python
# Questo genera un SyntaxError di proposito: manca la virgoletta di chiusura
print("Hello, Python!)
```

Cosa succede se un errore si verifica nel mezzo di più istruzioni? Tutto quello precedente viene eseguito, l'errore interrompe l'esecuzione, e ciò che segue **non arriva mai a essere eseguito**:

```python
print("Questo viene stampato")
frint("Questo genera un errore e blocca tutto ciò che segue")
print("Questo NON viene MAI stampato")
```

## 4. Variabili: salvare valori per usarli dopo

Una variabile è un contenitore con un nome. Si crea con `=`, che prende il valore a destra e lo salva nel nome a sinistra.

```python
# Salvare un valore in una variabile
x = 43 + 60 + 16 + 41
x
```

```python
# Usare una variabile per calcolarne un'altra
y = x / 60
y
```

```python
# Riassegnare: il valore precedente di x si perde e viene sostituito dal nuovo
x = x / 60
x
```

Scegliere buoni nomi di variabile aiuta a capire il codice senza dover indovinare. Confronta:

```python
# Nomi descrittivi: si capisce cosa rappresenta ogni valore
total_min = 43 + 42 + 57       # Durata totale in minuti
total_horas = total_min / 60   # Durata totale in ore
total_horas
```

## 5. Espressioni aritmetiche e ordine delle operazioni

Python supporta `+`, `-`, `*`, `/` (divisione) e inoltre `//` (**divisione intera**, arrotonda per difetto).

```python
# Operatori aritmetici di base
print(43 + 60 + 16 + 41)   # somma
print(50 - 60)              # sottrazione, può dare un risultato negativo
print(5 * 5)                # moltiplicazione
print(25 / 5)                # divisione
```

```python
# Divisione intera: arrotonda per difetto
print(25 // 5)   # divisione esatta
print(25 // 6)   # 25/6 = 4.166..., arrotondato per difetto -> 4

# Esempio pratico: quante ore complete ci sono in 160 minuti?
print(160 // 60)   # 2 ore complete (avanzano 40 minuti, che qui vengono scartati)
```

Python rispetta il consueto ordine matematico: moltiplicazione e divisione prima di somma e sottrazione, a meno che tu non usi le parentesi.

```python
# Senza parentesi: si risolve prima 2 * 60
print(30 + 2 * 60)     # 150

# Con parentesi: si risolve prima 30 + 2
print((30 + 2) * 60)   # 1920
```

```python
# Le parentesi sono fondamentali quando si combinano variabili in un'unica espressione
total_horas = (43 + 42 + 57) / 60  # prima si somma, poi si divide
total_horas
```

## 6. Tipi di dati: `int`, `float`, `str`, `bool` e casting

Ogni valore in Python ha un **tipo**, che determina quali operazioni gli si possono applicare:

- **`int`**: intero (`-1`, `0`, `100`)
- **`float`**: numero con decimali (`3.14`, `-42.0`)
- **`str`**: testo tra virgolette (`"Hello, Python 101!"`)
- **`bool`**: `True` o `False`

Usiamo `type()` per chiedere a Python di che tipo è un valore:

```python
print(type(12))                    # <class 'int'>
print(type(2.14))                  # <class 'float'>
print(type("Hello, Python 101!"))  # <class 'str'>
print(type(True))                  # <class 'bool'>
```

Attenzione: `1` e `1.0` **non sono la stessa cosa** per Python. La divisione `/` restituisce sempre un `float`, anche se il risultato sembra un numero intero:

```python
print(type(1.0))     # float, anche se sembra un intero
print(type(6 / 2))   # float, anche se 6/2 dà esattamente 3.0
print(type(6 // 2))  # int, perché // fa la divisione intera
```

**Casting** (conversione di tipi): `int()`, `float()`, `str()` e `bool()` convertono un valore in un altro tipo.

```python
# Da intero a float: non si perde informazione
print(float(2))       # 2.0

# Da float a intero: la parte decimale viene troncata (non arrotondata)
print(int(1.1))       # 1

# Da testo a numero, quando il testo rappresenta un numero valido
print(int('1'))       # 1
print(float('1.2'))   # 1.2

# Da numero a testo
print(str(1))         # '1'
print(str(1.2))       # '1.2'
```

Convertire testo in numero funziona solo se il testo ha *esattamente* la forma di un numero. Altrimenti, Python genera un errore. **Questa cella fallirà di proposito**:

```python
# Questo genera un errore di proposito: il testo non è un numero valido
int('1 or 2 people')
```

Anche i booleani si convertono da e verso i numeri: `True` equivale a `1`, `False` equivale a `0`; qualsiasi numero diverso da zero si converte in `True`, e `0` si converte in `False`.

```python
print(int(True))     # 1
print(int(False))    # 0
print(bool(1))        # True
print(bool(0))        # False
print(float(True))   # 1.0
```
