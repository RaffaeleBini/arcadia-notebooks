## ¿Por qué nos importa esto?

Hasta ahora escribiste bloques de código con `if`, `for` y `while` que resuelven un problema una vez. El problema aparece en cuanto necesitas ese mismo bloque en cinco sitios distintos del programa: copias y pegas, y el día que hay que corregir un error, tienes que acordarte de arreglarlo en los cinco sitios.

Una **función** te permite escribir esa lógica una sola vez, ponerle un nombre, y volver a usarla llamándola por ese nombre las veces que haga falta. Si hay un error, lo arreglas en un solo lugar.

En Ciencia de Datos vas a escribir funciones constantemente: una función que limpie una columna, una función que calcule una métrica, una función que reciba una URL y devuelva datos descargados. Y ya usaste funciones sin saberlo: `print()`, `len()`, `sum()`, `type()` son todas funciones — este capítulo te enseña a crear las tuyas.

## 1. Definir una función: `def`, `return` y docstrings

Piensa en una función como una receta de cocina: tiene un nombre, una lista de ingredientes (los **parámetros**) y un resultado final (lo que **`return`** entrega).

En Python defines una función con `def`, el nombre, paréntesis con los parámetros, y dos puntos. El cuerpo va indentado:

- **`def add(a):`** — el nombre es `add`, recibe un parámetro `a`.
- El **docstring** (`"""..."""` justo después de los dos puntos) documenta qué hace la función. Python lo usa para `help(nombre_funcion)`.
- **`return b`** — termina la función y "entrega" el valor de `b` a quien la llamó.

```python
def add(a):
    """
    Suma 1 a 'a' y devuelve el resultado.
    """
    b = a + 1
    print(a, "si le sumas uno, da", b)
    return b


add(1)
add(2)
```

Gracias al docstring, podemos pedirle ayuda a Python sobre esta función:

```python
help(add)
```

Un detalle importante: en cuanto Python ejecuta `return`, la función **termina ahí mismo**, aunque queden líneas de código después.

```python
def Mult(a, b):
    c = a * b
    return c
    print('Esto nunca se imprime')  # el return ya cortó la ejecución


resultado = Mult(12, 2)
print(resultado)  # 24
```

Si una función no tiene ningún `return`, Python le devuelve automáticamente `None`. Fíjate en la diferencia entre **imprimir** (mostrar algo en pantalla) y **devolver** (entregar un valor utilizable al código que llamó a la función):

```python
def MJ():
    print('Michael Jackson')


def MJ1():
    print('Michael Jackson')
    return None


print(MJ())    # imprime el texto, y luego imprime None (lo que devuelve MJ)
print(MJ1())   # exactamente lo mismo, return None es el comportamiento por defecto
```

Una función también puede combinar todo lo que ya conoces de estructuras de control: un `if` que decide qué devolver, o un `for` que recorre una lista que le pasas como parámetro.

```python
def type_of_album(artist, album, year_released):
    print(artist, album, year_released)
    if year_released > 1980:
        return "Moderno"
    else:
        return "Antiguo"


x = type_of_album("Michael Jackson", "Thriller", 1980)
print(x)  # Antiguo
```

```python
def PrintList(the_list):
    for element in the_list:
        print(element)


PrintList(['1', 1, 'the man', "abc"])
```

Un ejemplo más completo, que junta `def`, un bucle, un diccionario y `.split()`: contar cuántas veces aparece cada palabra en un texto.

```python
def freq(string):
    # Paso 1: separamos el texto en una lista de palabras
    words = string.split()

    # Paso 2: diccionario vacío para acumular los conteos
    Dict = {}

    # Paso 3: por cada palabra, contamos cuántas veces aparece en toda la lista
    for key in words:
        Dict[key] = words.count(key)

    print("La frecuencia de las palabras es:", Dict)


freq("Mary had a little lamb Little lamb, little lamb Mary had a little lamb")
```

Esta función junta prácticamente todo lo que viste en capítulos anteriores — strings, listas, diccionarios, bucles — dentro de un solo bloque reutilizable. Ese es justo el punto de una función: el lugar donde combinas piezas sueltas para resolver un problema concreto, y le pones un nombre para no tener que volver a pensarlo.

## 2. Parámetros y valores por defecto

Algunas recetas tienen ingredientes opcionales: si no dices nada, se usa una cantidad estándar. En Python esto es un **valor por defecto**, y se define directamente en la firma de la función:

```python
def isGoodRating(rating=4):
    if rating < 7:
        print("este disco es malo, su nota es", rating)
    else:
        print("este disco es bueno, su nota es", rating)


isGoodRating()      # usa rating=4 -> "este disco es malo, su nota es 4"
isGoodRating(10)     # usa rating=10 -> "este disco es bueno, su nota es 10"
```

Esto es útil cuando la mayoría de las veces vas a llamar a la función con el mismo valor típico, pero quieres dejar abierta la posibilidad de cambiarlo en casos puntuales.

⚠️ **Regla**: los parámetros con valor por defecto siempre van **después** de los que no lo tienen. `def f(a=1, b):` es un error en Python.

## 3. Scope: variables locales y globales

Esta es la parte donde más gente se confunde, así que vamos despacio. Piensa en cada función como una habitación con la puerta cerrada: lo que creas dentro (una **variable local**) se queda dentro y desaparece cuando la función termina.

```python
def square(a):
    b = 1              # b es una variable local: solo existe aquí dentro
    c = a * a + b
    print(a, "al cuadrado más 1 es", c)
    return c


x = 3
y = square(x)
print(y)     # 10 -> esto sí funciona, "y" es una variable normal fuera de la función
```

Si intentamos acceder a `b` fuera de la función, obtenemos un error, porque `b` nunca existió fuera de `square`:

```python
try:
    print(b)
except NameError as e:
    print("Error:", e)
```

En cambio, una **variable global** se define fuera de cualquier función, y es visible desde cualquier parte del programa, incluido el interior de las funciones (para *leerla*):

```python
myFavouriteBand = "AC/DC"   # variable global


def getBandRating(bandname):
    if bandname == myFavouriteBand:   # la función puede LEER la variable global
        return 10.0
    else:
        return 0.0


print("La nota de AC/DC es:", getBandRating("AC/DC"))
print("Mi banda favorita es:", myFavouriteBand)
```

Ahora la parte que más lía: si dentro de una función creas una variable con el **mismo nombre** que una global, Python **no toca la global**. Crea una variable local nueva que solo existe dentro de esa función, y "tapa" (hace *shadowing* de) a la global únicamente dentro de ese bloque.

```python
myFavouriteBand = "AC/DC"          # global


def getBandRating(bandname):
    myFavouriteBand = "Deep Purple"  # esto es una variable LOCAL nueva, no la de arriba
    if bandname == myFavouriteBand:
        return 10.0
    else:
        return 0.0


print("La nota de Deep Purple es:", getBandRating("Deep Purple"))  # 10.0
print("Mi banda favorita sigue siendo:", myFavouriteBand)          # AC/DC
```

¿Y si de verdad necesitas que una función **modifique** una variable global? Para eso existe la palabra clave **`global`**: le dice a Python "esta variable no es local, es la de fuera (o la quiero crear como global)".

```python
artist = "Michael Jackson"


def printer(artist):
    global internal_var
    internal_var = "Whitney Houston"
    print(artist, "es un artista")


printer(artist)
printer(internal_var)   # ahora internal_var sí existe fuera de la función
```

Sin ese `global`, `internal_var` habría sido una variable local que desaparece al terminar la función, y la segunda línea habría dado un `NameError`.

**Consejo honesto**: usa `global` con moderación. Es casi siempre mejor que una función reciba lo que necesita como parámetro y devuelva su resultado con `return`, en vez de leer y escribir variables globales por detrás.

## 4. `*args` y `**kwargs`: cuando no sabes cuántos argumentos vendrán

¿Qué haces si quieres una función que acepte "todos los argumentos que le pases", sin saber de antemano cuántos serán?

**`*args`** empaqueta todos los argumentos sueltos (posicionales) en una **tupla**:

```python
def printAll(*args):
    print("Número de argumentos:", len(args))
    for argument in args:
        print(argument)


printAll('Horsefeather', 'Adonis', 'Bone')
print()
printAll('Sidecar', 'Long Island', 'Mudslide', 'Carriage')  # funciona igual con 4 argumentos
```

El nombre `args` es solo convención (podrías llamarlo `*cualquier_cosa`); lo que le dice a Python "empaqueta aquí" es el asterisco `*`.

**`**kwargs`** hace lo mismo pero para argumentos `clave=valor`, empaquetados en un **diccionario**:

```python
def printDictionary(**args):
    for key in args:
        print(key + " : " + args[key])


printDictionary(Country='Canada', Province='Ontario', City='Toronto')
```

¿Cuándo usarías esto? Cuando escribes una función "genérica" que debe aceptar un número variable de valores sin anticipar todos los casos posibles — por ejemplo, una función que reciba cualquier cantidad de columnas de un dataset, o configuraciones opcionales con nombre. Las verás todo el tiempo en la documentación de librerías como `pandas`.

## 5. Pasar una lista a una función: cuidado con la mutabilidad

Cuando pasas un número o un string a una función, la función recibe una **copia** del valor. Pero las listas (y los diccionarios) funcionan distinto: cuando pasas una lista, pasas una **referencia** al mismo objeto en memoria. Si la función la modifica, el cambio se ve también fuera.

```python
def addItems(lista):
    lista.append("Three")
    lista.append("Four")


myList = ["One", "Two"]
addItems(myList)

print(myList)   # ['One', 'Two', 'Three', 'Four']
```

`addItems` no tiene ningún `return`, y aun así `myList` cambió fuera de la función. Esto pasa porque `lista` (el parámetro) y `myList` (la variable de fuera) apuntan **al mismo objeto**: no hay "dos listas", hay una sola lista con dos nombres apuntándole.

Esto es muy distinto de `square(a)`, donde `a` era un número: los números, strings y tuplas son **inmutables**, no se pueden modificar "en el sitio".

¿Por qué importa en Ciencia de Datos? Porque trabajarás constantemente con listas, diccionarios y, más adelante, DataFrames de `pandas`, que se comportan igual. Si una función modifica "por error" un objeto pensando que es una copia, puedes corromper datos originales. La solución, cuando necesitas que la función no toque el original, es hacer una copia explícita con `.copy()` antes de modificarla.

```python
def addItemsSeguro(lista):
    copia = lista.copy()   # trabajamos sobre una copia, no sobre el original
    copia.append("Three")
    copia.append("Four")
    return copia


myList2 = ["One", "Two"]
nuevaLista = addItemsSeguro(myList2)

print("Original (sin tocar):", myList2)
print("Nueva lista:", nuevaLista)
```

## Resumen: lo que te llevas de esta clase

- Una función es un bloque de código reutilizable: se define una vez con `def nombre(parámetros):` y se llama todas las veces que haga falta.
- `return` termina la función y entrega un valor; sin `return`, la función devuelve `None`. Imprimir con `print()` no es lo mismo que devolver con `return`.
- Puedes documentar una función con un docstring y consultarlo con `help(nombre_funcion)`.
- Una variable creada **dentro** de una función es **local**; una creada **fuera** de toda función es **global** y es visible en todo el programa, incluidas las funciones (para leerla).
- Si dentro de una función creas una variable con el mismo nombre que una global, se crea una variable local nueva que no toca a la global — a menos que uses `global nombre_variable`.
- `*args` empaqueta argumentos posicionales variables en una tupla; `**kwargs` empaqueta argumentos con nombre variables en un diccionario.
- Los parámetros con valor por defecto (`def f(x=4):`) siempre van después de los parámetros sin valor por defecto.
- Pasar una lista o un diccionario a una función pasa una **referencia**, no una copia: si la función la modifica, el cambio se ve también fuera. Usa `.copy()` cuando necesites evitarlo.
