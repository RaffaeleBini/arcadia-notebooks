## 1. ¿Por qué necesitamos listas y tuplas?

Hasta ahora guardábamos un solo valor por variable. Pero un álbum tiene varios datos asociados: artista, título, año, género, calificación. En vez de crear `album1_artista`, `album1_anio`, `album2_artista`... agrupamos todo en una sola estructura ordenada.

Una **secuencia ordenada** es una colección donde cada elemento tiene una posición fija (un índice), empezando en `0`. Listas y tuplas son las dos secuencias ordenadas básicas de Python. La diferencia entre ellas no está en cómo se accede a los datos, sino en **si se pueden modificar después de creadas**.

### 1.1 Listas — mutables, con corchetes `[ ]`

```python
# Una lista con datos mixtos sobre un álbum: artista, calificación, año
L = ["Michael Jackson", 10.1, 1982]
L
```

### 1.2 Tuplas — inmutables, con paréntesis `( )`

```python
# Una tupla con los mismos tipos de datos: string, entero, float
tuple1 = ("disco", 10, 1.2)
tuple1
```

```python
# El tipo de la variable es 'tuple'
type(tuple1)
```

Las listas se pueden modificar después de creadas; las tuplas, no. Si intentas cambiar un elemento de una tupla, Python lanza un `TypeError`.

```python
# Con una lista SI podemos reasignar una posición
A = ["disco", 10, 1.2]
print('Antes del cambio:', A)
A[0] = 'hard rock'
print('Después del cambio:', A)
```

```python
# Con una tupla, intentar lo mismo lanza un error
tupla_prueba = ("disco", 10, 1.2)
try:
    tupla_prueba[0] = 'hard rock'
except TypeError as e:
    print("¡Error! Las tuplas no se pueden modificar ->", e)
```

**¿Cuándo usar cada una?** Regla práctica: si los datos van a cambiar durante el programa, usa una **lista**. Si representan algo fijo que ya ocurrió o no debería tocarse, usa una **tupla**.

Una fila del dataset de álbumes (artista, año, título) describe un hecho histórico — buena candidata a tupla. Un carrito de compras que se va llenando y vaciando — buena candidata a lista. Además, al ser inmutables, las tuplas son ligeramente más rápidas y livianas en memoria.

## 2. Indexado y slicing

Funciona igual en listas y tuplas.

```python
# Indexado positivo: 0 es el primer elemento
L = ["Michael Jackson", 10.1, 1982]
print(L[0])
print(L[1])
print(L[2])
```

```python
# Indexado negativo: -1 es el último elemento, contando desde el final
print(L[-1])   # igual que L[2]
print(L[-2])   # igual que L[1]
print(L[-3])   # igual que L[0]
```

```python
# Exactamente igual con tuplas
tupla1 = ("disco", 10, 1.2)
print(tupla1[-1])
print(tupla1[-2])
```

**Slicing**: `secuencia[inicio:fin]` extrae un tramo. El elemento en la posición `fin` **no** se incluye.

```python
L = ["Michael Jackson", 10.1, 1982, "MJ", 1]
L[3:5]     # posiciones 3 y 4
```

```python
tupla2 = ("disco", 10, 1.2, "hard rock", 10)
print(tupla2[0:3])
print(tupla2[3:5])
print(len(tupla2))   # cantidad de elementos
```

El slicing de una lista devuelve una **lista nueva**; el de una tupla devuelve una **tupla nueva**. El tipo original se conserva.

Además, las tuplas se pueden **concatenar** con `+` (crea una tupla nueva, no modifica las originales) y se pueden **ordenar** con `sorted()` (que, como la tupla no se puede modificar, devuelve una **lista** nueva).

```python
# Concatenar tuplas
tupla1 = ("disco", 10, 1.2)
tupla2 = tupla1 + ("hard rock", 10)
tupla2
```

```python
# sorted() sobre una tupla devuelve una lista nueva y ordenada
Ratings = (0, 9, 6, 5, 10, 8, 9, 6, 2)
RatingsSorted = sorted(Ratings)
print(RatingsSorted, type(RatingsSorted))
```

## 3. Métodos principales de listas

| Método | Qué hace |
|---|---|
| `append(x)` | Agrega **un solo** elemento al final |
| `extend(iterable)` | Agrega **cada elemento** de un iterable, por separado |
| `insert(i, x)` | Inserta `x` en la posición `i` |
| `pop(i)` | Quita y devuelve el elemento en `i` (sin argumento: el último) |
| `remove(x)` | Quita la primera aparición del **valor** `x` |
| `sort()` | Ordena la lista en el lugar |
| `reverse()` | Invierte el orden en el lugar |
| `del lista[i]` | Elimina el elemento en la posición `i` (instrucción, no método) |

`append` vs `extend` es la confusión más común al empezar. `append` agrega el argumento **como un único bloque** (aunque sea una lista). `extend` **aplana** el iterable y agrega cada elemento suelto.

```python
# append: agrega UN elemento (que puede ser una lista completa)
L = ["Michael Jackson", 10.2]
L.append(["pop", 10])
L
```

```python
# extend: agrega CADA elemento por separado (aplana)
L = ["Michael Jackson", 10.2]
L.extend(["pop", 10])
L
```

```python
# insert, pop, remove, sort, del en acción
L = ["Michael Jackson", 10.2, "pop", 10]
L.insert(2, "nuevo genero")
print("Tras insert:", L)

elemento_quitado = L.pop()
print("Elemento quitado con pop():", elemento_quitado)
print("Lista tras pop():", L)

L.remove("nuevo genero")
print("Tras remove('nuevo genero'):", L)

A = ["hard rock", 10, 1.2]
print('Antes de del:', A)
del A[0]
print('Después de del:', A)
```

```python
# sort() ordena en el lugar; sorted() devuelve una lista nueva
numeros = [5, 2, 9, 1]
numeros.sort()
print("sort() en el lugar:", numeros)

letras = ["c", "a", "b"]
letras_ordenadas = sorted(letras)
print("sorted() -> nueva lista:", letras_ordenadas, "| original intacta:", letras)
```

`split()` conecta strings con listas: convierte un string en una lista, separando por espacios (por defecto) o por el delimitador que le indiques. Muy útil al leer datos crudos, por ejemplo una línea de un CSV.

```python
print('hard rock'.split())
print('Michael Jackson,Thriller,1982'.split(','))
```

## 4. Listas y tuplas anidadas (nesting)

Tanto listas como tuplas pueden contener, como cualquier otro elemento, otra lista o tupla. Cada índice adicional te baja un nivel más — como abrir subcarpetas dentro de una carpeta.

```python
# Una lista puede mezclar tipos, incluyendo otra lista y una tupla
["Michael Jackson", 10.1, 1982, [1, 2], ("A", 1)]
```

```python
NestedT = (1, 2, ("pop", "rock"), (3, 4), ("disco", (1, 2)))

print("NestedT[2]:      ", NestedT[2])       # ('pop', 'rock')
print("NestedT[2][0]:   ", NestedT[2][0])    # 'pop'
print("NestedT[2][1]:   ", NestedT[2][1])    # 'rock'
print("NestedT[4]:      ", NestedT[4])       # ('disco', (1, 2))
print("NestedT[4][1]:   ", NestedT[4][1])    # (1, 2)
print("NestedT[4][1][0]:", NestedT[4][1][0]) # 1  -> tres niveles de profundidad
print("NestedT[4][1][1]:", NestedT[4][1][1]) # 2
```

## 5. Copia por referencia vs. clonado — la trampa más común

Cuando escribes `B = A` con listas, **no** creas una segunda lista independiente. Creas una segunda etiqueta que apunta al **mismo objeto** en memoria (copia por referencia). Si modificas `A`, el cambio también se ve en `B`.

```python
A = ["hard rock", 10, 1.2]
B = A   # copia por referencia: misma lista, dos nombres
print('A:', A)
print('B:', B)
```

```python
print('B[0] antes:', B[0])
A[0] = "banana"
print('B[0] después de cambiar A:', B[0])   # ¡cambió, aunque solo tocamos A!
```

Para obtener una copia **de verdad**, independiente, usamos el slicing completo `[:]` (o el método `.copy()`). Esto se llama **clonar**.

```python
A = ["banana", 10, 1.2]
B = A[:]   # clonado: una lista NUEVA con los mismos valores
print('B[0] antes:', B[0])
A[0] = "hard rock"
print('B[0] después de cambiar A:', B[0])   # B no cambió, es independiente
```

Este problema **no existe con tuplas**: al ser inmutables, `B = A` nunca puede morderte, porque ninguna de las dos etiquetas puede cambiar el contenido de todos modos. Una razón práctica más para preferir tuplas cuando los datos no necesitan cambiar.

## 6. Aplicando todo: el dataset de álbumes

Volvamos al escenario inicial: recomendaciones de álbumes de tus amigos. Cada álbum es una **tupla** (dato fijo, ya ocurrió), y la colección de recomendaciones es una **lista** (puede crecer con nuevas recomendaciones).

```python
# Cada álbum: (Artista, Título, Año, Género, Calificación de amigos)
album_thriller = ("Michael Jackson", "Thriller", 1982, "Pop, rock, R&B", 10.0)
album_back_in_black = ("AC/DC", "Back in Black", 1980, "Hard rock", 8.5)
album_dark_side = ("Pink Floyd", "The Dark Side of the Moon", 1973, "Progressive rock", 9.5)

# La lista de recomendaciones SI puede crecer con el tiempo
recomendaciones = [album_thriller, album_back_in_black]
print("Recomendaciones iniciales:")
for album in recomendaciones:
    print(" -", album[1], "de", album[0], f"({album[2]})")

# Llega una nueva recomendación
recomendaciones.append(album_dark_side)
print("\nDespués de la nueva recomendación:")
for album in recomendaciones:
    print(" -", album[1], "de", album[0], f"({album[2]})")

# Intentar modificar un dato del álbum (una tupla) falla, como debe ser
try:
    album_thriller[2] = 1983
except TypeError as e:
    print("\nNo se puede modificar un hecho histórico:", e)
```

## Resumen de la clase

- **Secuencia ordenada**: colección con posiciones fijas, accesibles por índice desde `0`. Listas y tuplas son las dos básicas.
- **Listas** (`[ ]`) son **mutables**; **tuplas** (`( )`) son **inmutables**.
- Usa listas para datos que cambian; tuplas para datos fijos (un hecho, una fila de registro histórico).
- Indexado (`L[0]`, `L[-1]`) y slicing (`L[inicio:fin]`) funcionan igual en ambas; el slicing devuelve una secuencia nueva del mismo tipo.
- `append()` agrega un solo bloque; `extend()` aplana y agrega cada elemento. Es la confusión más común.
- `insert`, `pop`, `remove`, `sort`, `reverse` modifican la lista **en el lugar**; `sorted()` devuelve una lista nueva sin tocar la original.
- Listas y tuplas se pueden **anidar**: cada índice adicional baja un nivel (`L[2][0][1]`).
- `B = A` con listas **no copia**, crea una segunda etiqueta al mismo objeto. Para clonar de verdad: `B = A[:]` o `A.copy()`.
- Este problema de referencia **no existe en tuplas**, porque no se pueden modificar de todos modos.
