## Perché questo ci interessa?

Le liste di Python sono flessibili ma lente per il calcolo numerico: sommare due liste di un milione di numeri elemento per elemento richiede un ciclo `for`, e Python controlla il tipo di ogni elemento a ogni passo. **NumPy** (Numerical Python) risolve questo problema con un nuovo tipo di dato, l'`ndarray`, che conserva dati dello stesso tipo in memoria compatta e applica le operazioni a tutto l'array in un colpo solo — senza che tu debba scrivere il ciclo. Questo si chiama **vettorizzazione**.

Non è solo una questione di velocità: **pandas**, la libreria che userai nel prossimo capitolo per lavorare con le tabelle, è costruita sopra NumPy. Ogni colonna di un DataFrame di pandas è, sotto il cofano, un array di NumPy.

```python
# NumPy e matplotlib vengono preinstallati su Colab e nella maggior parte delle installazioni di Jupyter
import numpy as np
import matplotlib.pyplot as plt

print(np.__version__)
```

## 1. Array 1D: creazione e `dtype`

Un array 1D è ciò che in matematica si chiama **vettore**. Si crea a partire da una lista con `np.array()`. Siccome tutti gli elementi condividono lo stesso tipo, l'array ha un attributo `dtype` che lo indica. Se mescoli interi e decimali, NumPy converte tutto in `float64`.

```python
# Creazione di un array 1D a partire da una lista
a = np.array([0, 1, 2, 3, 4])
print(a)
print(type(a))
print(a.dtype)   # int64

b = np.array([3.1, 11.02, 6.2, 213.2, 5.2])
print(b.dtype)   # float64 -> ci sono decimali, NumPy unifica il tipo
```

## 2. Indicizzazione, assegnazione e slicing

Accedere e modificare elementi funziona come nelle liste di Python: parentesi quadre per accedere, `array[inizio:fine:passo]` per lo slicing (l'indice `fine` resta escluso).

```python
a = np.array([0, 1, 2, 3, 4])
print(a[0], a[4])

# Riassegnare un valore singolo
c = np.array([20, 1, 2, 3, 4])
c[0] = 100
print(c)

# Slicing: array[inizio:fine] -> l'indice fine non è incluso
d = c[1:4]
print(d)

# Slicing con passo: array[inizio:fine:passo]
arr = np.array([1, 2, 3, 4, 5, 6, 7])
print(arr[1:5:2])   # dall'indice 1 al 5, di 2 in 2
print(arr[:4])       # se ometti l'inizio, parte da 0
print(arr[4:])       # se ometti la fine, arriva fino all'ultimo

# Assegnare un intero tratto in un colpo solo
c[3:5] = 300, 400
print(c)
```

Una cosa che le liste di Python non fanno in modo così diretto: selezionare più indici sparsi passando **un'altra lista** come "lista di indici" dentro le parentesi quadre.

```python
c = np.array([100, 1, 2, 300, 400])

select = [0, 2, 3, 4]
print(c[select])

# Quella stessa selezione serve anche per riassegnare valori in blocco
c[select] = 100000
print(c)
```

## 3. Attributi e statistiche

`size`, `ndim` e `shape` descrivono la forma dell'array (in 1D non dicono ancora molto, ma sono fondamentali in 2D). I metodi `mean()`, `std()`, `max()` e `min()` calcolano statistiche senza che tu debba scrivere la formula.

```python
a = np.array([0, 1, 2, 3, 4])
print("size:", a.size)    # numero totale di elementi
print("ndim:", a.ndim)    # numero di dimensioni
print("shape:", a.shape)  # dimensione in ciascun asse, come tupla

datos = np.array([1, -1, 1, -1])
print("media:", datos.mean())
print("deviazione standard:", datos.std())

b = np.array([-1, 2, 3, 4, 5])
print("massimo:", b.max())
print("minimo:", b.min())
```

## 4. Operazioni aritmetiche vettorizzate

Con le liste di Python, sommare due collezioni elemento per elemento richiede un ciclo `for`. Con gli array di NumPy, l'operatore `+` sa già che vuoi sommare posizione contro posizione: è un'**operazione vettorizzata**. Tutte le operazioni di base funzionano così, e NumPy offre inoltre le versioni come funzione (`np.add`, `np.subtract`, `np.divide`), equivalenti agli operatori.

```python
# Confronto: con le liste pure serve un ciclo
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
resultado_lista = [lista1[i] + lista2[i] for i in range(len(lista1))]
print("con liste:", resultado_lista)

# Con gli array di NumPy, l'operatore vettorizza già l'operazione
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
print("con array:", array1 + array2)

u = np.array([1, 0])
v = np.array([0, 1])
z = np.add(u, v)   # equivalente a u + v
print("np.add:", z)

a_ = np.array([10, 20, 30])
b_ = np.array([5, 10, 15])
print("np.subtract:", np.subtract(a_, b_))
print("np.divide:", np.divide(a_, b_))
```

Puoi anche operare un array con un unico numero (uno **scalare**): NumPy applica l'operazione a ogni elemento. E quando moltiplichi due array con `*`, NumPy moltiplica posizione contro posizione — questo si chiama **prodotto di Hadamard** (moltiplicazione elemento per elemento) e **non è la stessa cosa** del prodotto scalare che vedremo più avanti.

```python
array = np.array([1, 2, 3])
print(array * 2)   # moltiplicazione per scalare -> [2 4 6]
print(array + 1)   # somma con scalare -> [2 3 4]

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
resultado = array1 * array2   # Hadamard: 1*4, 2*5, 3*6
print("Hadamard:", resultado)
```

## 5. Visualizzare i vettori e il prodotto scalare

Prima di calcolare il prodotto scalare con il codice, vediamo i vettori come frecce. La funzione seguente, usando `matplotlib`, disegna due vettori e la loro somma.

```python
def plot_vectores(u, z, v):
    ax = plt.axes()
    ax.arrow(0, 0, *u, head_width=0.05, color='r', head_length=0.1)
    plt.text(*(u + 0.1), 'u')

    ax.arrow(0, 0, *v, head_width=0.05, color='b', head_length=0.1)
    plt.text(*(v + 0.1), 'v')

    ax.arrow(0, 0, *z, head_width=0.05, head_length=0.1)
    plt.text(*(z + 0.1), 'z')
    plt.ylim(-2, 2)
    plt.xlim(-2, 2)

u = np.array([1, 0])
v = np.array([0, 1])
z = np.add(u, v)

plot_vectores(u, z, v)
plt.show()
```

L'intuizione del **prodotto scalare**: immagina di aver venduto 3 mele, 2 pere e 5 banane (quantità), a prezzi di 1.5, 2.0 e 0.8. Il ricavo totale non si ottiene sommando quantità e prezzi: **li accoppi per posizione, moltiplichi e sommi tutto**. Questo è esattamente `np.dot()` — a differenza di Hadamard, non ti restituisce un nuovo array, ma un unico numero che "riassume" la combinazione dei due vettori.

```python
cantidades = np.array([3, 2, 5])
precios = np.array([1.5, 2.0, 0.8])

total = np.dot(cantidades, precios)
print("ricavo totale:", total)   # 12.5

X = np.array([1, 2])
Y = np.array([3, 2])
print("np.dot(X, Y):", np.dot(X, Y))   # 1*3 + 2*2 = 7

# Confronta con Hadamard: stessa coppia di array, risultato molto diverso
print("Hadamard X*Y:", X * Y)      # [3, 4] -> un array
print("Dot np.dot(X,Y):", np.dot(X, Y))  # 7    -> un unico numero
```

## 6. Funzioni matematiche e `np.linspace()`

NumPy porta con sé funzioni matematiche che si applicano elemento per elemento (`np.sin`, ecc.). `np.linspace(inizio, fine, num=n)` genera `n` numeri distribuiti in modo uniforme tra `inizio` e `fine` — molto utile per graficare funzioni matematiche.

```python
# 100 numeri equidistanti tra 0 e 2*pi
x = np.linspace(0, 2 * np.pi, num=100)
y = np.sin(x)   # il seno di ciascuno di questi 100 numeri

plt.plot(x, y)
plt.title("y = sin(x)")
plt.show()
```

## 7. Array 2D: creazione e indicizzazione

Un array 2D è una **matrice**: una tabella organizzata in righe e colonne. Si costruisce passando una lista di liste, dove ogni lista interna è una riga. Ora sì che `ndim`, `shape` e `size` mostrano la loro utilità.

```python
lista_de_listas = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
A = np.array(lista_de_listas)
print(A)

print("ndim:", A.ndim)    # 2 -> righe e colonne
print("shape:", A.shape)  # (3, 3) -> 3 righe, 3 colonne
print("size:", A.size)    # 9 elementi in totale
```

Per accedere a un elemento servono due coordinate: riga e colonna. `A[riga, colonna]` è la forma idiomatica di NumPy (internamente più diretta di `A[riga][colonna]`, che prima crea una riga intermedia).

```python
A = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])

print(A[1, 2])   # riga indice 1, colonna indice 2 -> 23
print(A[1][2])   # stesso risultato, altra sintassi

# Slicing in 2D: la virgola separa righe e colonne
print(A[0][0:2])   # prima riga, colonne 0 e 1  -> [11 12]
print(A[0:2, 2])   # righe 0 e 1, colonna 2        -> [13 23]
print(A[:, 1])      # tutte le righe, colonna 1    -> [12 22 32]
print(A[1])          # tutta la riga indice 1         -> [21 22 23]
```

## 8. Operazioni con le matrici: somma, scalare e Hadamard

Le stesse regole degli array 1D si estendono al 2D, purché le matrici abbiano la stessa forma (`shape`).

```python
X = np.array([[1, 0], [0, 1]])
Y = np.array([[2, 1], [1, 2]])

# Somma di matrici: posizione contro posizione
Z = X + Y
print("somma:\n", Z)

# Moltiplicazione per scalare: ogni elemento per il numero
Z = 2 * Y
print("scalare:\n", Z)

# Prodotto di Hadamard: elemento per elemento
Z = X * Y
print("Hadamard:\n", Z)
```

## 9. Moltiplicazione tra matrici con `np.dot()` e trasposizione

`np.dot()` su matrici 2D estende l'idea del prodotto scalare: ogni riga della prima matrice contro ogni colonna della seconda. Regola di compatibilità: una matrice **m × n** si può moltiplicare solo per una **n × p**, e il risultato è **m × p**. L'ordine conta — la moltiplicazione tra matrici non è commutativa.

```python
A = np.array([[0, 1, 1], [1, 0, 1]])      # forma (2, 3)
B = np.array([[1, 1], [1, 1], [-1, 1]])   # forma (3, 2)

Z = np.dot(A, B)
print("shape del risultato:", Z.shape)   # (2, 2) -> il 3 in mezzo si semplifica
print(Z)

# Verifica manuale della prima casella:
# riga 0 di A = [0, 1, 1], colonna 0 di B = [1, 1, -1]
# 0*1 + 1*1 + 1*(-1) = 0
print("Z[0,0] a mano:", 0*1 + 1*1 + 1*(-1))
```

La **trasposizione** ruota la matrice: le righe diventano colonne e viceversa. Si ottiene con l'attributo `.T`.

```python
C = np.array([[1, 1], [2, 2], [3, 3]])
print("C shape:", C.shape)   # (3, 2)
print(C)

print("C.T shape:", C.T.shape)   # (2, 3)
print(C.T)
```

## Riepilogo della lezione

- NumPy esiste perché le liste di Python sono lente per il calcolo numerico su larga scala; gli array sono strutture uniformi per tipo e dimensione che permettono operazioni **vettorizzate**, senza cicli `for` espliciti.
- Un array 1D è un vettore, un array 2D è una matrice. `ndim`, `shape` e `size` descrivono le dimensioni, la grandezza per ciascun asse e il numero totale di elementi.
- L'indicizzazione e lo slicing funzionano come nelle liste; in 2D si usa una virgola: `A[riga, colonna]`.
- Le operazioni `+`, `-`, `*`, `/` tra array sono elemento per elemento. `array1 * array2` è il prodotto di **Hadamard**, non il prodotto scalare.
- Il prodotto scalare (`np.dot()` in 1D) combina due vettori e restituisce **un unico numero**. La moltiplicazione tra matrici (`np.dot()` in 2D) estende questa idea a righe contro colonne, con la regola **m×n · n×p = m×p**, e l'ordine conta.
- `.T` traspone una matrice (righe ↔ colonne).
- pandas è costruito sopra NumPy: tutto ciò che hai visto qui riappare in ogni colonna di un DataFrame.
