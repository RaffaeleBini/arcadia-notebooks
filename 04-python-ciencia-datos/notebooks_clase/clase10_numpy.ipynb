## ¿Por qué nos importa esto?

Las listas de Python son flexibles pero lentas para cálculo numérico: sumar dos listas de un millón de números elemento a elemento exige un bucle `for`, y Python revisa el tipo de cada elemento en cada paso. **NumPy** (Numerical Python) resuelve esto con un tipo de dato nuevo, el `ndarray`, que guarda datos del mismo tipo en memoria compacta y aplica operaciones a todo el array de golpe — sin que tú escribas el bucle. Esto se llama **vectorización**.

Esto no es solo una cuestión de velocidad: **pandas**, la librería que vas a usar en el próximo capítulo para trabajar con tablas, está construida encima de NumPy. Cada columna de un DataFrame de pandas es, por debajo, un array de NumPy.

```python
# NumPy y matplotlib vienen preinstalados en Colab y en la mayoría de instalaciones de Jupyter
import numpy as np
import matplotlib.pyplot as plt

print(np.__version__)
```

## 1. Arrays 1D: creación y `dtype`

Un array 1D es lo que en matemáticas se llama **vector**. Se crea a partir de una lista con `np.array()`. Como todos los elementos comparten tipo, el array tiene un atributo `dtype` que lo indica. Si mezclas enteros y decimales, NumPy convierte todo a `float64`.

```python
# Creando un array 1D a partir de una lista
a = np.array([0, 1, 2, 3, 4])
print(a)
print(type(a))
print(a.dtype)   # int64

b = np.array([3.1, 11.02, 6.2, 213.2, 5.2])
print(b.dtype)   # float64 -> hay decimales, NumPy unifica el tipo
```

## 2. Indexado, asignación y slicing

Acceder y modificar elementos funciona igual que en las listas de Python: corchetes para acceder, `array[inicio:fin:paso]` para slicing (el índice `fin` queda excluido).

```python
a = np.array([0, 1, 2, 3, 4])
print(a[0], a[4])

# Reasignar un valor individual
c = np.array([20, 1, 2, 3, 4])
c[0] = 100
print(c)

# Slicing: array[inicio:fin] -> el índice fin no se incluye
d = c[1:4]
print(d)

# Slicing con paso: array[inicio:fin:paso]
arr = np.array([1, 2, 3, 4, 5, 6, 7])
print(arr[1:5:2])   # desde el índice 1 hasta el 5, de 2 en 2
print(arr[:4])       # si omites el inicio, arranca en 0
print(arr[4:])       # si omites el fin, llega hasta el final

# Asignar un tramo entero de una sola vez
c[3:5] = 300, 400
print(c)
```

Algo que las listas de Python no hacen tan directamente: seleccionar varios índices sueltos pasando **otra lista** como "lista de índices" dentro de los corchetes.

```python
c = np.array([100, 1, 2, 300, 400])

select = [0, 2, 3, 4]
print(c[select])

# Esa misma selección sirve para reasignar valores de golpe
c[select] = 100000
print(c)
```

## 3. Atributos y estadísticas

`size`, `ndim` y `shape` describen la forma del array (con 1D no aportan mucho todavía, pero son claves en 2D). Los métodos `mean()`, `std()`, `max()` y `min()` calculan estadísticas sin que escribas la fórmula.

```python
a = np.array([0, 1, 2, 3, 4])
print("size:", a.size)    # número total de elementos
print("ndim:", a.ndim)    # número de dimensiones
print("shape:", a.shape)  # tamaño en cada dimensión, como tupla

datos = np.array([1, -1, 1, -1])
print("media:", datos.mean())
print("desviación estándar:", datos.std())

b = np.array([-1, 2, 3, 4, 5])
print("máximo:", b.max())
print("mínimo:", b.min())
```

## 4. Operaciones aritméticas vectorizadas

Con listas de Python, sumar dos colecciones elemento a elemento exige un bucle `for`. Con arrays de NumPy, el operador `+` ya sabe que quieres sumar posición contra posición: es una **operación vectorizada**. Todas las operaciones básicas funcionan así, y NumPy además ofrece las versiones como función (`np.add`, `np.subtract`, `np.divide`), equivalentes a los operadores.

```python
# Comparación: con listas puras necesitas un bucle
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
resultado_lista = [lista1[i] + lista2[i] for i in range(len(lista1))]
print("con listas:", resultado_lista)

# Con arrays de NumPy, el operador ya vectoriza la operación
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
print("con arrays:", array1 + array2)

u = np.array([1, 0])
v = np.array([0, 1])
z = np.add(u, v)   # equivalente a u + v
print("np.add:", z)

a_ = np.array([10, 20, 30])
b_ = np.array([5, 10, 15])
print("np.subtract:", np.subtract(a_, b_))
print("np.divide:", np.divide(a_, b_))
```

También puedes operar un array contra un único número (un **escalar**): NumPy aplica la operación a cada elemento. Y cuando multiplicas dos arrays con `*`, NumPy multiplica posición contra posición — esto se llama **producto Hadamard** (multiplicación elemento a elemento) y **no es lo mismo** que el producto punto que viene más abajo.

```python
array = np.array([1, 2, 3])
print(array * 2)   # multiplicación escalar -> [2 4 6]
print(array + 1)   # suma escalar -> [2 3 4]

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
resultado = array1 * array2   # Hadamard: 1*4, 2*5, 3*6
print("Hadamard:", resultado)
```

## 5. Visualizando vectores y el producto punto

Antes de calcular el producto punto con código, veamos los vectores como flechas. La siguiente función, usando `matplotlib`, dibuja dos vectores y su suma.

```python
def plot_vectores(u, z, v):
    ax = plt.axes()
    ax.arrow(0, 0, *u, head_width=0.05, color='r', head_length=0.1)
    plt.text(*(u + 0.1), 'u')

    ax.arrow(0, 0, *v, head_width=0.05, color='b', head_length=0.1)
    plt.text(*(v + 0.1), 'v')

    ax.arrow(0, 0, *z, head_width=0.05, head_length=0.1)
    plt.text(*(z + 0.1), 'z')
    plt.ylim(-2, 2)
    plt.xlim(-2, 2)

u = np.array([1, 0])
v = np.array([0, 1])
z = np.add(u, v)

plot_vectores(u, z, v)
plt.show()
```

La intuición del **producto punto**: imagina que vendiste 3 manzanas, 2 peras y 5 plátanos (cantidades), a precios de 1.5, 2.0 y 0.8. El ingreso total no es sumar cantidades con precios: **emparejas por posición, multiplicas y sumas todo**. Eso es exactamente `np.dot()` — a diferencia del Hadamard, no te devuelve un array nuevo, sino un único número que "resume" la combinación de ambos vectores.

```python
cantidades = np.array([3, 2, 5])
precios = np.array([1.5, 2.0, 0.8])

total = np.dot(cantidades, precios)
print("ingreso total:", total)   # 12.5

X = np.array([1, 2])
Y = np.array([3, 2])
print("np.dot(X, Y):", np.dot(X, Y))   # 1*3 + 2*2 = 7

# Compara con Hadamard: misma pareja de arrays, resultado muy distinto
print("Hadamard X*Y:", X * Y)      # [3, 4] -> un array
print("Dot np.dot(X,Y):", np.dot(X, Y))  # 7    -> un solo número
```

## 6. Funciones matemáticas y `np.linspace()`

NumPy trae funciones matemáticas que se aplican elemento a elemento (`np.sin`, etc.). `np.linspace(inicio, fin, num=n)` genera `n` números repartidos de forma uniforme entre `inicio` y `fin` — muy útil para graficar funciones matemáticas.

```python
# 100 números igualmente espaciados entre 0 y 2*pi
x = np.linspace(0, 2 * np.pi, num=100)
y = np.sin(x)   # el seno de cada uno de esos 100 números

plt.plot(x, y)
plt.title("y = sin(x)")
plt.show()
```

## 7. Arrays 2D: creación e indexado

Un array 2D es una **matriz**: una tabla organizada en filas y columnas. Se construye pasando una lista de listas, donde cada lista interna es una fila. Ahora sí, `ndim`, `shape` y `size` muestran su utilidad.

```python
lista_de_listas = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
A = np.array(lista_de_listas)
print(A)

print("ndim:", A.ndim)    # 2 -> filas y columnas
print("shape:", A.shape)  # (3, 3) -> 3 filas, 3 columnas
print("size:", A.size)    # 9 elementos en total
```

Para acceder a un elemento necesitas dos coordenadas: fila y columna. `A[fila, columna]` es la forma idiomática de NumPy (más directa internamente que `A[fila][columna]`, que primero crea una fila intermedia).

```python
A = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])

print(A[1, 2])   # fila índice 1, columna índice 2 -> 23
print(A[1][2])   # mismo resultado, otra sintaxis

# Slicing en 2D: coma separa filas y columnas
print(A[0][0:2])   # primera fila, columnas 0 y 1  -> [11 12]
print(A[0:2, 2])   # filas 0 y 1, columna 2        -> [13 23]
print(A[:, 1])      # todas las filas, columna 1    -> [12 22 32]
print(A[1])          # toda la fila índice 1         -> [21 22 23]
```

## 8. Operaciones con matrices: suma, escalar y Hadamard

Las mismas reglas de los arrays 1D se extienden a 2D, siempre que las matrices tengan la misma forma (`shape`).

```python
X = np.array([[1, 0], [0, 1]])
Y = np.array([[2, 1], [1, 2]])

# Suma de matrices: posición contra posición
Z = X + Y
print("suma:\n", Z)

# Multiplicación por escalar: cada elemento por el número
Z = 2 * Y
print("escalar:\n", Z)

# Producto Hadamard: elemento a elemento
Z = X * Y
print("Hadamard:\n", Z)
```

## 9. Multiplicación de matrices con `np.dot()` y transposición

`np.dot()` sobre matrices 2D extiende la idea del producto punto: cada fila de la primera matriz contra cada columna de la segunda. Regla de compatibilidad: una matriz **m × n** solo se puede multiplicar por una **n × p**, y el resultado es **m × p**. El orden importa — la multiplicación de matrices no es conmutativa.

```python
A = np.array([[0, 1, 1], [1, 0, 1]])      # forma (2, 3)
B = np.array([[1, 1], [1, 1], [-1, 1]])   # forma (3, 2)

Z = np.dot(A, B)
print("shape del resultado:", Z.shape)   # (2, 2) -> el 3 del medio se cancela
print(Z)

# Verificación manual de la primera casilla:
# fila 0 de A = [0, 1, 1], columna 0 de B = [1, 1, -1]
# 0*1 + 1*1 + 1*(-1) = 0
print("Z[0,0] a mano:", 0*1 + 1*1 + 1*(-1))
```

La **transposición** gira la matriz: las filas pasan a ser columnas y viceversa. Se obtiene con el atributo `.T`.

```python
C = np.array([[1, 1], [2, 2], [3, 3]])
print("C shape:", C.shape)   # (3, 2)
print(C)

print("C.T shape:", C.T.shape)   # (2, 3)
print(C.T)
```

## Resumen de la clase

- NumPy existe porque las listas de Python son lentas para cálculo numérico a gran escala; los arrays son estructuras uniformes en tipo y tamaño que permiten operaciones **vectorizadas**, sin bucles `for` explícitos.
- Un array 1D es un vector, un array 2D es una matriz. `ndim`, `shape` y `size` describen dimensiones, tamaño por dimensión y número total de elementos.
- El indexado y el slicing funcionan como en listas; en 2D se usa una coma: `A[fila, columna]`.
- Las operaciones `+`, `-`, `*`, `/` entre arrays son elemento a elemento. `array1 * array2` es el producto **Hadamard**, no el producto punto.
- El producto punto (`np.dot()` en 1D) combina dos vectores y devuelve **un único número**. La multiplicación de matrices (`np.dot()` en 2D) extiende esa idea a filas contra columnas, con la regla **m×n · n×p = m×p**, y el orden importa.
- `.T` transpone una matriz (filas ↔ columnas).
- pandas está construido sobre NumPy: todo lo visto aquí reaparece en cada columna de un DataFrame.
