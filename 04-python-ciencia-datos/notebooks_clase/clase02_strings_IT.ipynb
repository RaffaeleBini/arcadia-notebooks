## 1. Cos'è una stringa?

Una stringa (`str`) è testo tra virgolette — semplici o doppie, per Python non fa differenza. Può contenere lettere, cifre come testo, spazi e caratteri speciali.

L'idea chiave per tutto ciò che segue: **una stringa è una sequenza ordinata di caratteri**, ognuno con una posizione fissa. Questa posizione è ciò che sfrutteremo nella prossima sezione.


```python
# Virgolette doppie o semplici: è indifferente
"The BodyGuard"
'The BodyGuard'

# Cifre e caratteri speciali restano testo
'1 2 3 4 5 6 '
'@#2_#]&*^%$'

# Stampare e salvare in variabile
print("hello!")

name = "The BodyGuard"
name
```

## 2. Indicizzazione positiva e negativa

Pensa a una stringa come a un treno di vagoni: ogni vagone è un carattere, e ha un numero di posto. **La numerazione inizia da 0**, non da 1.

- `name[0]` → primo carattere
- `name[6]` → settimo carattere (posizione 6)

Python permette anche di contare **dalla fine** con indici negativi: `-1` è l'ultimo carattere, `-2` il penultimo, ecc.

Per sapere quanti caratteri ha una stringa, usiamo `len(...)`.


```python
name = "The BodyGuard"

# Indicizzazione positiva: si conta da 0
print(name[0])    # 'T', primo carattere
print(name[6])    # 'o'
print(name[10])   # 'u'

# Indicizzazione negativa: si conta dalla fine, iniziando da -1
print(name[-1])    # 'd', ultimo carattere
print(name[-13])   # 'T', primo carattere (la stringa ha 13 caratteri)

# len() conta i caratteri, spazi inclusi
print(len(name))   # 13
```

## 3. Slicing con stride: ritagliare pezzi di testo

L'indicizzazione ti dà **un** carattere; lo slicing ti dà un **pezzo**. Sintassi: `string[inizio:fine]`.

Trabocchetto da memorizzare: l'indice di `fine` **non è incluso**. `name[0:4]` restituisce le posizioni 0, 1, 2 e 3 — quattro caratteri.

Puoi aggiungere un terzo numero, lo *stride* (passo): `string[inizio:fine:passo]`. Se lasci vuoti inizio e fine (`::2`), Python assume "dall'inizio alla fine".


```python
name = "The BodyGuard"

# Slicing di base: fine NON è incluso
print(name[0:4])     # 'The '
print(name[8:12])    # 'Guar'

# Slicing con stride (passo)
print(name[::2])      # 'TeBdGad' -> un carattere sì, uno no, per tutta la stringa
print(name[0:5:2])    # 'TeB'    -> primi 5 caratteri, saltandone uno ogni volta
```

### Concatenare e ripetere stringhe

Due operazioni che userai continuamente: `+` per unire stringhe, `*` per ripeterle.

Attenzione: le stringhe sono **immutabili**. `name + "..."` non modifica `name` sul posto — crea una stringa nuova. Se vuoi "farla crescere", devi riassegnarla.


```python
name = "The BodyGuard"

# Concatenare con +
statement = name + " is the best album"
print(statement)

# Ripetere con *
print(3 * "The BodyGuard")

# Pattern comune: riassegnare alla stessa variabile
name = name + " is the best album"
print(name)
```

## 4. Sequenze di escape e raw string

Una barra rovesciata (`\`) seguita da una lettera dice a Python "questo non è testo letterale, è un carattere speciale". Le due più comuni: `\n` (a capo) e `\t` (tabulazione). Per mostrare una barra rovesciata vera e propria, si esegue l'escape su se stessa: `\\`.

Questo diventa un problema con i percorsi di Windows (`C:\new_folder\file.txt`), perché Python interpreta `\n` e `\f` come escape e rovina il percorso. La soluzione: una **raw string**, anteponendo `r` prima delle virgolette — dice a Python "non interpretare nulla come escape".

**Regola pratica:** se la tua stringa contiene barre rovesciate e non vuoi che vengano interpretate (percorsi di Windows, pattern RegEx), usa `r"..."`.


```python
# \n = a capo, \t = tabulazione
print(" The BodyGuard\n is the best album")
print(" The BodyGuard \t is the best album")

# Barra rovesciata letterale: fa l'escape su se stessa
print(" The BodyGuard \\ is the best album")

# Problema classico: percorsi di Windows senza raw string
regular_string = "C:\new_folder\file.txt"
print(regular_string)   # si rompe: \n e \f vengono interpretati come escape

# Soluzione: raw string
raw_string = r"C:\new_folder\file.txt"
print(raw_string)
```

## 5. Metodi per manipolare le stringhe

Un metodo è una funzione "attaccata" al valore: `string.metodo()`. Nessuno modifica la stringa originale (sono immutabili) — tutti ne restituiscono una **nuova**.


### `.upper()` e `.lower()`

Utili per confrontare testo senza che importi come lo ha scritto ognuno (`"Rock"`, `"rock"`, `"ROCK"` dovrebbero essere la stessa categoria).


```python
a = "Thriller is the sixth studio album"
print("before upper:", a)

b = a.upper()
print("After upper:", b)

# .lower() fa l'opposto
print(b.lower())
```

### `.replace()`

Cerca un pezzo di testo e lo sostituisce con un altro, restituendo una stringa nuova. Serve per correggere errori di battitura, unificare categorie, o cancellare testo (sostituendolo con `""`).


```python
a = "The BodyGuard is the best album"
b = a.replace('BodyGuard', 'Janet')
print(b)
```

### `.find()`

Restituisce la posizione (indice) in cui **inizia** la prima occorrenza di ciò che cerchi. Se non la trova, restituisce `-1` (non genera un errore).


```python
name = "The BodyGuard"

print(name.find('he'))      # 1
print(name.find('Guard'))   # 8
print(name.find('Jasdfasdasdf'))   # -1: non l'ha trovato

# Uso tipico: verifica rapida di presenza
texto = "si è verificato un errore al passo 3"
if texto.find("errore") != -1:
    print("Trovata la parola 'errore'")
```

### `.split()`

Divide la stringa in base al separatore indicato (per default, qualsiasi spazio bianco) e restituisce una **lista** dei pezzi.


```python
name = "The BodyGuard"
split_string = name.split()
print(split_string)

# Con separatore esplicito e limite di tagli
csv_row = "Whitney Houston,1985,Pop"
print(csv_row.split(","))
print(csv_row.split(",", 1))   # maxsplit=1: al massimo un taglio
```

## 6. Formattare stringhe: comporre testo con dei dati

Ora il caso inverso: hai dei dati separati e vuoi comporli in una frase leggibile. Python ha tre modi — li vedrai tutti perché compaiono in codice reale, ma ce n'è uno preferito: le **f-string**.


### f-string (consigliato)

Da Python 3.6 in poi: anteponi `f` alle virgolette e scrivi variabili (o intere espressioni) tra `{}`.


```python
name = "John"
age = 30
print(f"My name is {name} and I am {age} years old.")

# Puoi anche mettere un'espressione completa dentro {}
x = 10
y = 20
print(f"The sum of x and y is {x+y}.")
```

### `.format()` (precedente alle f-string)

Colloca i valori nell'ordine in cui compaiono le parentesi graffe vuote `{}`. Lo vedrai in molto codice esistente.


```python
name = "John"
age = 50
print("My name is {} and I am {} years old.".format(name, age))
```

### L'operatore `%` (il più vecchio)

Usa segnaposto come `%s` (testo) e `%d` (intero), e una tupla di valori dopo `%`. Lo riconoscerai in codice datato, ma oggi non conviene adottarlo.


```python
name = "Johnathan"
age = 30
print("My name is %s and I am %d years old." % (name, age))
```

## 7. Introduzione alle RegEx: quando cercare testo esatto non basta

`.find()` e `.replace()` cercano testo **letterale**. Molte volte ti serve trovare un **pattern** ("dieci cifre di seguito", "qualsiasi parola tra virgolette"), non un testo fisso. Per questo c'è il modulo `re`, incluso in Python — basta importarlo.

| Sequenza | Significa |
|---|---|
| `\d` | Qualsiasi cifra (0-9) |
| `\D` | Qualsiasi carattere che NON sia una cifra |
| `\w` | Carattere di parola (lettere, cifre, `_`) |
| `\W` | Carattere che NON è di parola |
| `\s` | Spazio bianco (spazio, tab, a capo) |
| `\S` | Carattere che NON è spazio bianco |

**Convenzione:** il pattern si scrive quasi sempre come raw string (`r"..."`), affinché Python non interpreti la barra rovesciata prima che arrivi a `re`.


### `re.search()`: il pattern esiste da qualche parte nel testo?

Restituisce un oggetto di corrispondenza (vero in un `if`) o `None` se non lo trova. `match.group()` restituisce il testo esatto che ha corrisposto.


```python
import re

s1 = "The BodyGuard is the best album"
pattern = r"Body"

result = re.search(pattern, s1)
if result:
    print("Match found!")
else:
    print("Match not found.")

# Un pattern vero: un numero di telefono di dieci cifre
pattern = r"\d\d\d\d\d\d\d\d\d\d"
text = "My Phone number is 1234567890"
match = re.search(pattern, text)

if match:
    print("Phone number found:", match.group())
else:
    print("No match")
```

### `re.findall()`: tutte le corrispondenze, non solo la prima


```python
pattern = r"\W"   # qualsiasi carattere che non sia di parola
text = "Hello, world!"
matches = re.findall(pattern, text)
print("Matches:", matches)

s2 = "The BodyGuard is the best album of 'Whitney Houston'."
result = re.findall("st", s2)
print(result)
```

### `re.split()`: dividere il testo usando un pattern come separatore


```python
s2 = "The BodyGuard is the best album of 'Whitney Houston'."
split_array = re.split(r"\s", s2)
print(split_array)
```

### `re.sub()`: sostituire per pattern

Come `.replace()`, ma cercando per pattern, con opzioni come `flags=re.IGNORECASE`.


```python
pattern = r"Whitney Houston"
replacement = "legend"

new_string = re.sub(pattern, replacement, s2, flags=re.IGNORECASE)
print(new_string)
```

## Riepilogo: cosa porti a casa da questa lezione

- Una stringa è una sequenza ordinata di caratteri; si accede per posizione (indice), a partire da 0. Gli indici negativi contano dalla fine (`-1` è l'ultimo).
- Lo slicing (`string[inizio:fine:passo]`) ritaglia un pezzo della stringa; l'indice di `fine` **non è mai** incluso.
- Le stringhe sono **immutabili**: nessun metodo le modifica sul posto.
- `\n`, `\t`, `\\` sono sequenze di escape; le raw string (`r"..."`) le disattivano — indispensabili con i percorsi di Windows e i pattern RegEx.
- `.upper()`, `.lower()`, `.replace()`, `.find()` e `.split()` coprono la maggior parte delle attività quotidiane di pulizia del testo.
- Per comporre testo con dei dati, preferisci le **f-string** rispetto a `.format()` e a `%`.
- Il modulo `re` entra in gioco quando il pattern che cerchi varia, non è un testo esatto: `re.search()`, `re.findall()`, `re.split()` e `re.sub()`.
