## ¿Por qué nos importa esto?

Hasta ahora tu código se ejecuta siempre igual, de arriba hacia abajo. Pero trabajar con datos reales significa **tomar decisiones** ("¿este valor es válido?") y **repetir procesos** ("haz esto con cada fila"). Condicionales y bucles son las dos piezas que te permiten hacer ambas cosas, y casi siempre se usan juntas: recorres una colección con un bucle y, dentro, decides con un `if` qué hacer con cada elemento.

## 1. Operadores de comparación

| Operador | Significado |
|---|---|
| `==` | igual a |
| `!=` | distinto de |
| `>` | mayor que |
| `<` | menor que |
| `>=` | mayor o igual que |
| `<=` | menor o igual que |

⚠️ **Ojo**: `=` asigna un valor, `==` compara dos valores. Es el error más común al empezar.

```python
edad = 25
print(edad == 25)   # ¿edad vale 25?
print(edad != 30)   # ¿edad es distinto de 30?
print(edad >= 20)   # ¿edad es mayor o igual a 20?
```

Estos operadores también funcionan con **strings**. Python compara según el código ASCII de cada carácter (las mayúsculas valen menos que las minúsculas), y la comparación es sensible a mayúsculas/minúsculas.

```python
print("ACDC" == "Michael Jackson")   # False, son strings distintos
print("ACDC" != "Michael Jackson")   # True

print('B' > 'A')     # True -> 66 > 65 en ASCII
print('BA' > 'AB')   # True -> compara primero la primera letra ('B' > 'A')
```

## 2. `if` / `else`: tomar decisiones

Piensa en un `if` como una puerta con llave: si la condición es `True`, la puerta se abre y el bloque indentado se ejecuta; si es `False`, se ignora por completo, sin generar ningún error.

```python
edad = 19

if edad > 18:
    print("puedes entrar al concierto")

print("continúa el programa")   # esto se ejecuta siempre, con puerta abierta o cerrada
```

`else` cubre el caso en que la condición del `if` fue falsa. No lleva condición propia: es el "en cualquier otro caso".

```python
edad = 17

if edad > 18:
    print("puedes entrar al concierto de rock")
else:
    print("mejor ve a ver la película")
```

### `elif`: más de dos caminos

Cuando hay más de dos posibilidades, encadenas condiciones con `elif` ("else if"). Python las evalúa **en orden** y ejecuta el primer bloque verdadero; ignora el resto, aunque también se cumplieran.

```python
edad = 18

if edad > 18:
    print("vas al concierto A (ACDC)")
elif edad == 18:
    print("vas al concierto B (Pink Floyd)")   # esta es la que se ejecuta
else:
    print("vas al concierto C (Meat Loaf)")

print("sigues con tu día")
```

Un uso muy típico de `elif` encadenados: clasificar por rangos, como el desayuno de un comedor escolar según la edad del estudiante.

```python
edad = 10

if edad <= 9:
    print("recibe papilla de avena")
elif edad >= 10 and edad <= 14:
    print("recibe un sándwich")
elif edad >= 15 and edad <= 17:
    print("recibe una hamburguesa")
```

### Condicionales anidados

Puedes poner un `if` dentro de otro `if` (por ejemplo, para validar un detalle dentro de una rama principal, como en un cajero automático). Es perfectamente válido, pero si te encuentras anidando 4 o 5 niveles, probablemente puedas simplificarlo combinando condiciones con operadores lógicos — el siguiente tema.

```python
opcion = "Retirar efectivo"

if opcion == "Retirar efectivo":
    monto = 50
    if monto % 10 == 0:
        print("Monto entregado:", monto)
    else:
        print("Ingresa un múltiplo de 10.")
else:
    print("Gracias por usar el cajero.")
```

## 3. Operadores lógicos: `and`, `or`, `not`

- **`and`** — todas las condiciones deben cumplirse (como una cerradura que exige tarjeta *y* huella dactilar).
- **`or`** — basta con que una condición se cumpla (como elegir película entre amigos: si a alguien le gusta el género, se elige).
- **`not`** — invierte el resultado de una condición.

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
tiene_tarjeta = True
tiene_huella = True

if tiene_tarjeta and tiene_huella:
    print("acceso concedido")

le_gusta_comedia = True
le_gusta_accion = False
le_gusta_drama = False

if le_gusta_comedia or le_gusta_accion or le_gusta_drama:
    print("se elige una película")

no_molestar = True

if not no_molestar:
    print("enviar notificación")
else:
    print("no se envía notificación, modo no molestar activo")
```

Un uso muy frecuente en ciencia de datos es comprobar si un valor cae **dentro** o **fuera** de un rango, combinando dos comparaciones con `and` u `or`.

```python
anio_album = 1985

if anio_album > 1979 and anio_album < 1990:
    print("el álbum es de los 80")

anio_album = 1995

if anio_album < 1980 or anio_album > 1989:
    print("el álbum no es de los 80")
else:
    print("el álbum es de los 80")
```

## 4. El bucle `for`: repetir sobre una secuencia

Usa `for` cuando **sabes de antemano** sobre qué secuencia vas a iterar (una lista, un string, un `range()`) y quieres procesar cada uno de sus elementos.

```python
colores = ["rojo", "naranja", "amarillo", "verde", "azul", "índigo", "violeta"]

for color in colores:
    print(color)
```

### `range()`: generar secuencias de números

- `range(n)` genera números desde `0` hasta `n - 1`.
- `range(inicio, fin)` genera números desde `inicio` hasta `fin - 1` (el límite superior nunca se incluye).

```python
for numero in range(11):
    print(numero)   # imprime del 0 al 10

print("---")

for numero in range(1, 11):
    print(numero)   # imprime del 1 al 10
```

`range(len(lista))` sirve para recorrer una lista **por índice** en vez de por valor, útil cuando necesitas ese índice para algo más — por ejemplo, para modificar la lista mientras la recorres.

```python
cuadrados = ['rojo', 'amarillo', 'verde', 'morado', 'azul']

for i in range(len(cuadrados)):
    print("antes:", cuadrados[i])
    cuadrados[i] = 'blanco'
    print("después:", cuadrados[i])
```

### Ejemplo real: tablas de multiplicar

Un uso clásico de `for` + `range()`: generar tablas de multiplicar completas sin escribir cada línea a mano.

```python
print("Tabla del 6:")
for i in range(10):
    print("6 *", i, "=", 6 * i)

print("\nTabla del 7:")
for i in range(10):
    print("7 *", i, "=", 7 * i)
```

### `enumerate()`: el índice y el valor a la vez

Si solo necesitas el índice para saber "en qué posición estoy" pero también quieres el valor, `enumerate()` te entrega ambos en cada vuelta y es más legible ("pythónico") que llevar tú mismo la cuenta con un contador.

```python
frutas = ["manzana", "banana", "naranja"]

for indice, fruta in enumerate(frutas):
    print(f"en la posición {indice} está {fruta}")
```

## 5. El bucle `while`: repetir mientras se cumpla una condición

Usa `while` cuando **no sabes de antemano** cuántas vueltas vas a necesitar: repites una tarea mientras una condición siga siendo verdadera.

```python
contador = 1
while contador <= 10:
    print(contador)
    contador += 1   # sin esto, la condición nunca cambia y el bucle no termina nunca
```

Un ejemplo más realista: recorrer una lista de años y detenerte apenas encuentres uno en particular. Aquí `while` es la herramienta correcta porque no sabes de antemano en qué posición está 1973 — la condición de parada depende del contenido de los datos, no de un conteo fijo.

```python
anios = [1982, 1980, 1973, 2000]

i = 0
anio_actual = anios[0]

while anio_actual != 1973:
    print(anio_actual)
    i += 1
    anio_actual = anios[i]

print("tomó", i, "vueltas encontrar 1973")
```

### Ejemplo real: filtrar animales por longitud de nombre

Otro caso muy típico: recorrer una colección y construir una lista nueva solo con los elementos que cumplen una condición.

```python
animales = ["leon", "jirafa", "gorila", "loros", "cocodrilo", "ciervo", "cisne"]
animales_siete_letras = []

i = 0
while i < len(animales):
    if len(animales[i]) == 7:
        animales_siete_letras.append(animales[i])
    i += 1

print(animales_siete_letras)
```

## 6. `for` vs. `while`: ¿cuál elijo?

- **Usa `for`** cuando conoces de antemano la secuencia sobre la que vas a iterar (una lista completa, un rango, cada carácter de un string). El número de vueltas está determinado por el tamaño de la secuencia.
- **Usa `while`** cuando el número de repeticiones depende de una condición que solo se conoce en tiempo de ejecución ("hasta que encuentres el valor", "mientras el rating no baje de 6").

Regla práctica: si puedes contestar "voy a repetir esto exactamente N veces, o una vez por cada elemento de esta colección", usa `for`. Si tu respuesta es "voy a repetir esto hasta que pase algo", usa `while` — y asegúrate siempre de que "ese algo" tenga manera de ocurrir, o el bucle nunca terminará.

```python
# La misma tarea resuelta con for y con while, para comparar

# Sumar los primeros 5 números con for (sabemos exactamente cuántas vueltas)
suma_for = 0
for n in range(1, 6):
    suma_for += n
print("suma con for:", suma_for)

# La misma suma con while (funciona, pero for es más directo aquí)
suma_while = 0
n = 1
while n <= 5:
    suma_while += n
    n += 1
print("suma con while:", suma_while)
```

## Resumen de la clase

- `==`, `!=`, `>`, `<`, `>=`, `<=` devuelven `True`/`False` y son la base de cualquier decisión. No confundas `==` (comparar) con `=` (asignar).
- `if` ejecuta un bloque solo si su condición es verdadera; `else` cubre todos los demás casos; `elif` encadena más condiciones y Python ejecuta la primera que se cumple.
- La indentación define qué código pertenece a un bloque `if`/`elif`/`else` — no es un detalle estético, es sintaxis.
- `and` exige que todas las condiciones sean verdaderas, `or` exige que al menos una lo sea, `not` invierte el resultado.
- `for` recorre una secuencia conocida; `range(n)` genera números de `0` a `n-1`; `enumerate()` te da índice y valor a la vez.
- `while` repite mientras una condición sea verdadera; usa cuando no sabes de antemano cuántas repeticiones harán falta — y nunca olvides actualizar la variable de la condición, o tendrás un bucle infinito.
- Condicionales y bucles casi siempre se combinan: recorres una colección con un bucle y, dentro, decides con un `if` qué hacer con cada elemento — el patrón que vas a usar para limpiar y filtrar datos en los próximos capítulos.
