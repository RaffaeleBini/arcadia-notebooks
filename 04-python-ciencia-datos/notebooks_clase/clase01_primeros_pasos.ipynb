## 3. Errores comunes: `NameError` y `SyntaxError`

Como Python ejecuta línea por línea, si hay un error en una línea, todo lo anterior ya se ejecutó normalmente y la ejecución se detiene justo ahí.

- **`NameError`**: usaste un nombre (función o variable) que Python no reconoce.
- **`SyntaxError`**: rompiste una regla "gramatical" del lenguaje (por ejemplo, una comilla sin cerrar).

Veamos un `NameError` en acción — escribimos mal `print` como `frint`. **Esta celda va a fallar a propósito**, es el objetivo del ejemplo:

```python
# Esto genera un NameError a propósito (frint no existe)
frint("Hello, Python!")
```

Ahora un `SyntaxError` — nos olvidamos de cerrar la comilla. **Esta celda también va a fallar a propósito**:

```python
# Esto genera un SyntaxError a propósito: falta la comilla de cierre
print("Hello, Python!)
```

¿Qué pasa si un error ocurre en medio de varias instrucciones? Todo lo anterior se ejecuta, el error corta la ejecución, y lo que sigue **nunca llega a correr**:

```python
print("Esto se imprime")
frint("Esto genera un error y detiene todo lo que sigue")
print("Esto NUNCA se imprime")
```

## 4. Variables: guardar valores para usarlos después

Una variable es un contenedor con nombre. Se crea con `=`, que toma el valor de la derecha y lo guarda en el nombre de la izquierda.

```python
# Guardar un valor en una variable
x = 43 + 60 + 16 + 41
x
```

```python
# Usar una variable para calcular otra
y = x / 60
y
```

```python
# Reasignar: el valor anterior de x se pierde y queda reemplazado por el nuevo
x = x / 60
x
```

Elegir buenos nombres de variable ayuda a que el código se entienda sin adivinar. Compará:

```python
# Nombres descriptivos: se entiende qué representa cada valor
total_min = 43 + 42 + 57       # Duración total en minutos
total_horas = total_min / 60   # Duración total en horas
total_horas
```

## 5. Expresiones aritméticas y orden de las operaciones

Python soporta `+`, `-`, `*`, `/` (división) y además `//` (**división entera**, redondea hacia abajo).

```python
# Operadores aritméticos básicos
print(43 + 60 + 16 + 41)   # suma
print(50 - 60)              # resta, puede dar negativo
print(5 * 5)                # multiplicación
print(25 / 5)                # división
```

```python
# División entera: redondea hacia abajo
print(25 // 5)   # división exacta
print(25 // 6)   # 25/6 = 4.166..., redondeado hacia abajo -> 4

# Ejemplo práctico: ¿cuántas horas completas hay en 160 minutos?
print(160 // 60)   # 2 horas completas (sobran 40 minutos, que acá se descartan)
```

Python respeta el orden matemático habitual: multiplicación y división antes que suma y resta, salvo que uses paréntesis.

```python
# Sin paréntesis: primero se resuelve 2 * 60
print(30 + 2 * 60)     # 150

# Con paréntesis: primero se resuelve 30 + 2
print((30 + 2) * 60)   # 1920
```

```python
# Los paréntesis son claves al combinar variables en una sola expresión
total_horas = (43 + 42 + 57) / 60  # se suma primero, se divide después
total_horas
```

## 6. Tipos de datos: `int`, `float`, `str`, `bool` y casting

Todo valor en Python tiene un **tipo**, que determina qué operaciones se le pueden aplicar:

- **`int`**: entero (`-1`, `0`, `100`)
- **`float`**: número con decimales (`3.14`, `-42.0`)
- **`str`**: texto entre comillas (`"Hello, Python 101!"`)
- **`bool`**: `True` o `False`

Usamos `type()` para preguntarle a Python de qué tipo es un valor:

```python
print(type(12))                    # <class 'int'>
print(type(2.14))                  # <class 'float'>
print(type("Hello, Python 101!"))  # <class 'str'>
print(type(True))                  # <class 'bool'>
```

Ojo: `1` y `1.0` **no son lo mismo** para Python. La división `/` siempre devuelve `float`, aunque el resultado parezca un número entero:

```python
print(type(1.0))     # float, aunque parezca un entero
print(type(6 / 2))   # float, aunque 6/2 dé exactamente 3.0
print(type(6 // 2))  # int, porque // hace división entera
```

**Casting** (conversión de tipos): `int()`, `float()`, `str()` y `bool()` convierten un valor a otro tipo.

```python
# De entero a float: no se pierde información
print(float(2))       # 2.0

# De float a entero: se trunca la parte decimal (no se redondea)
print(int(1.1))       # 1

# De texto a número, cuando el texto representa un número válido
print(int('1'))       # 1
print(float('1.2'))   # 1.2

# De número a texto
print(str(1))         # '1'
print(str(1.2))       # '1.2'
```

Convertir texto a número solo funciona si el texto tiene *exactamente* la forma de un número. Si no, Python lanza un error. **Esta celda va a fallar a propósito**:

```python
# Esto genera un error a propósito: el texto no es un número válido
int('1 or 2 people')
```

Los booleanos también se convierten hacia y desde números: `True` equivale a `1`, `False` equivale a `0`; cualquier número distinto de cero se convierte en `True`, y `0` se convierte en `False`.

```python
print(int(True))     # 1
print(int(False))    # 0
print(bool(1))        # True
print(bool(0))        # False
print(float(True))   # 1.0
```
