## 1. ¿Qué es un string?

Un string (`str`) es texto entre comillas — simples o dobles, Python no hace diferencia. Puede contener letras, dígitos como texto, espacios y caracteres especiales.

La idea clave para todo lo que sigue: **un string es una secuencia ordenada de caracteres**, cada uno con una posición fija. Esa posición es lo que vamos a explotar en la próxima sección.


```python
# Comillas dobles o simples: da lo mismo
"The BodyGuard"
'The BodyGuard'

# Dígitos y caracteres especiales siguen siendo texto
'1 2 3 4 5 6 '
'@#2_#]&*^%$'

# Imprimir y guardar en variable
print("hello!")

name = "The BodyGuard"
name
```

## 2. Indexado positivo y negativo

Pensá un string como un tren de vagones: cada vagón es un carácter, y tiene un número de asiento. **La numeración arranca en 0**, no en 1.

- `name[0]` → primer carácter
- `name[6]` → séptimo carácter (posición 6)

Python también permite contar **desde el final** con índices negativos: `-1` es el último carácter, `-2` el anteúltimo, etc.

Para saber cuántos caracteres tiene un string, usamos `len(...)`.


```python
name = "The BodyGuard"

# Indexado positivo: se cuenta desde 0
print(name[0])    # 'T', primer carácter
print(name[6])    # 'o'
print(name[10])   # 'u'

# Indexado negativo: se cuenta desde el final, empezando en -1
print(name[-1])    # 'd', último carácter
print(name[-13])   # 'T', primer carácter (el string tiene 13 caracteres)

# len() cuenta caracteres, incluyendo espacios
print(len(name))   # 13
```

## 3. Slicing con stride: recortar pedazos de texto

El indexado te da **un** carácter; el slicing te da un **pedazo**. Sintaxis: `string[inicio:fin]`.

Trampa a memorizar: el índice de `fin` **no se incluye**. `name[0:4]` da las posiciones 0, 1, 2 y 3 — cuatro caracteres.

Podés agregar un tercer número, el *stride* (paso): `string[inicio:fin:paso]`. Si dejás vacíos inicio y fin (`::2`), Python asume "de principio a fin".


```python
name = "The BodyGuard"

# Slicing básico: fin NO se incluye
print(name[0:4])     # 'The '
print(name[8:12])    # 'Guar'

# Slicing con stride (paso)
print(name[::2])      # 'TeBdGad' -> un caracter sí, uno no, todo el string
print(name[0:5:2])    # 'TeB'    -> primeros 5 caracteres, salteando de a uno
```

### Concatenar y repetir strings

Dos operaciones que vas a usar todo el tiempo: `+` para pegar strings, `*` para repetirlos.

Ojo: los strings son **inmutables**. `name + "..."` no modifica `name` in situ — crea un string nuevo. Si querés "hacerlo crecer", tenés que reasignarlo.


```python
name = "The BodyGuard"

# Concatenar con +
statement = name + " is the best album"
print(statement)

# Repetir con *
print(3 * "The BodyGuard")

# Patrón común: reasignar a la misma variable
name = name + " is the best album"
print(name)
```

## 4. Secuencias de escape y raw strings

Una barra invertida (`\`) seguida de una letra le dice a Python "esto no es texto literal, es un carácter especial". Las dos más comunes: `\n` (salto de línea) y `\t` (tabulador). Para mostrar una barra invertida real, se escapa a sí misma: `\\`.

Esto se vuelve un problema con rutas de Windows (`C:\new_folder\file.txt`), porque Python interpreta `\n` y `\f` como escapes y arruina la ruta. La solución: un **raw string**, anteponiendo `r` antes de las comillas — le dice a Python "no interpretes nada como escape".

**Regla práctica:** si tu string tiene barras invertidas y no querés que se interpreten (rutas de Windows, patrones de RegEx), usá `r"..."`.


```python
# \n = salto de línea, \t = tabulador
print(" The BodyGuard\n is the best album")
print(" The BodyGuard \t is the best album")

# Barra invertida literal: se escapa a sí misma
print(" The BodyGuard \\ is the best album")

# Problema clásico: rutas de Windows sin raw string
regular_string = "C:\new_folder\file.txt"
print(regular_string)   # se rompe: \n y \f se interpretan como escapes

# Solución: raw string
raw_string = r"C:\new_folder\file.txt"
print(raw_string)
```

## 5. Métodos para manipular strings

Un método es una función "pegada" al valor: `string.metodo()`. Ninguno modifica el string original (son inmutables) — todos devuelven uno **nuevo**.


### `.upper()` y `.lower()`

Útiles para comparar texto sin que importe cómo lo escribió cada quien (`"Rock"`, `"rock"`, `"ROCK"` deberían ser la misma categoría).


```python
a = "Thriller is the sixth studio album"
print("before upper:", a)

b = a.upper()
print("After upper:", b)

# .lower() hace lo opuesto
print(b.lower())
```

### `.replace()`

Busca un pedazo del texto y lo cambia por otro, devolviendo un string nuevo. Sirve para arreglar errores de tipeo, unificar categorías, o borrar texto (reemplazando por `""`).


```python
a = "The BodyGuard is the best album"
b = a.replace('BodyGuard', 'Janet')
print(b)
```

### `.find()`

Devuelve la posición (índice) donde **empieza** la primera aparición de lo que buscás. Si no lo encuentra, devuelve `-1` (no tira error).


```python
name = "The BodyGuard"

print(name.find('he'))      # 1
print(name.find('Guard'))   # 8
print(name.find('Jasdfasdasdf'))   # -1: no lo encontró

# Uso típico: chequeo rápido de presencia
texto = "se produjo un error en el paso 3"
if texto.find("error") != -1:
    print("Se encontró la palabra 'error'")
```

### `.split()`

Corta el string por el separador indicado (por defecto, cualquier espacio en blanco) y devuelve una **lista** de los pedazos.


```python
name = "The BodyGuard"
split_string = name.split()
print(split_string)

# Con separador explícito y límite de cortes
csv_row = "Whitney Houston,1985,Pop"
print(csv_row.split(","))
print(csv_row.split(",", 1))   # maxsplit=1: como máximo un corte
```

## 6. Formatear strings: armar texto con datos

Ahora el caso inverso: tenés datos sueltos y querés armarlos en una oración legible. Python tiene tres formas — vas a ver las tres porque aparecen en código real, pero hay una preferida: **f-strings**.


### f-strings (recomendado)

Desde Python 3.6: anteponé `f` a las comillas y escribí variables (o expresiones enteras) entre `{}`.


```python
name = "John"
age = 30
print(f"My name is {name} and I am {age} years old.")

# También podés poner una expresión completa dentro de {}
x = 10
y = 20
print(f"The sum of x and y is {x+y}.")
```

### `.format()` (anterior a f-strings)

Coloca los valores en el orden en que aparecen las llaves vacías `{}`. Lo vas a ver en mucho código existente.


```python
name = "John"
age = 50
print("My name is {} and I am {} years old.".format(name, age))
```

### El operador `%` (el más antiguo)

Usa marcadores como `%s` (texto) y `%d` (entero), y una tupla de valores después de `%`. Lo vas a reconocer en código antiguo, pero no conviene adoptarlo hoy.


```python
name = "Johnathan"
age = 30
print("My name is %s and I am %d years old." % (name, age))
```

## 7. Introducción a RegEx: cuando buscar texto exacto no alcanza

`.find()` y `.replace()` buscan texto **literal**. Muchas veces necesitás encontrar un **patrón** ("diez dígitos seguidos", "cualquier palabra entre comillas"), no un texto fijo. Para eso está el módulo `re`, incluido en Python — solo hace falta importarlo.

| Secuencia | Significa |
|---|---|
| `\d` | Cualquier dígito (0-9) |
| `\D` | Cualquier carácter que NO sea dígito |
| `\w` | Carácter de palabra (letras, dígitos, `_`) |
| `\W` | Carácter que NO es de palabra |
| `\s` | Espacio en blanco (espacio, tab, salto de línea) |
| `\S` | Carácter que NO es espacio en blanco |

**Convención:** el patrón casi siempre se escribe como raw string (`r"..."`), para que Python no interprete la barra invertida antes de que llegue a `re`.


### `re.search()`: ¿existe el patrón en algún lado del texto?

Devuelve un objeto de coincidencia (verdadero en un `if`) o `None` si no lo encuentra. `match.group()` devuelve el texto exacto que matcheó.


```python
import re

s1 = "The BodyGuard is the best album"
pattern = r"Body"

result = re.search(pattern, s1)
if result:
    print("Match found!")
else:
    print("Match not found.")

# Un patrón de verdad: un teléfono de diez dígitos
pattern = r"\d\d\d\d\d\d\d\d\d\d"
text = "My Phone number is 1234567890"
match = re.search(pattern, text)

if match:
    print("Phone number found:", match.group())
else:
    print("No match")
```

### `re.findall()`: todas las coincidencias, no solo la primera


```python
pattern = r"\W"   # cualquier carácter que no sea de palabra
text = "Hello, world!"
matches = re.findall(pattern, text)
print("Matches:", matches)

s2 = "The BodyGuard is the best album of 'Whitney Houston'."
result = re.findall("st", s2)
print(result)
```

### `re.split()`: partir el texto usando un patrón como separador


```python
s2 = "The BodyGuard is the best album of 'Whitney Houston'."
split_array = re.split(r"\s", s2)
print(split_array)
```

### `re.sub()`: reemplazar por patrón

Como `.replace()`, pero buscando por patrón, con opciones como `flags=re.IGNORECASE`.


```python
pattern = r"Whitney Houston"
replacement = "legend"

new_string = re.sub(pattern, replacement, s2, flags=re.IGNORECASE)
print(new_string)
```

## Resumen: lo que te llevas de esta clase

- Un string es una secuencia ordenada de caracteres; se accede por posición (índice), empezando en 0. Los índices negativos cuentan desde el final (`-1` es el último).
- El slicing (`string[inicio:fin:paso]`) recorta un pedazo del string; el índice de `fin` **nunca** se incluye.
- Los strings son **inmutables**: ningún método los modifica in situ.
- `\n`, `\t`, `\\` son secuencias de escape; los raw strings (`r"..."`) las desactivan — imprescindibles con rutas de Windows y patrones de RegEx.
- `.upper()`, `.lower()`, `.replace()`, `.find()` y `.split()` cubren la mayoría de las tareas de limpieza de texto del día a día.
- Para armar texto con datos, preferí **f-strings** sobre `.format()` y sobre `%`.
- El módulo `re` entra en juego cuando el patrón que buscás varía, no un texto exacto: `re.search()`, `re.findall()`, `re.split()` y `re.sub()`.
