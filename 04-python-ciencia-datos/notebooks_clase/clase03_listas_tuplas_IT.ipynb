## 1. Perché abbiamo bisogno di liste e tuple?

Finora salvavamo un solo valore per variabile. Ma un album ha diversi dati associati: artista, titolo, anno, genere, valutazione. Invece di creare `album1_artista`, `album1_anio`, `album2_artista`... raggruppiamo tutto in un'unica struttura ordinata.

Una **sequenza ordinata** è una collezione in cui ogni elemento ha una posizione fissa (un indice), a partire da `0`. Liste e tuple sono le due sequenze ordinate di base in Python. La differenza tra loro non sta in come si accede ai dati, ma nel **fatto che si possano modificare dopo essere state create**.

### 1.1 Liste — mutabili, con parentesi quadre `[ ]`

```python
# Una lista con dati misti su un album: artista, valutazione, anno
L = ["Michael Jackson", 10.1, 1982]
L
```

### 1.2 Tuple — immutabili, con parentesi tonde `( )`

```python
# Una tupla con gli stessi tipi di dati: stringa, intero, float
tuple1 = ("disco", 10, 1.2)
tuple1
```

```python
# Il tipo della variabile è 'tuple'
type(tuple1)
```

Le liste si possono modificare dopo essere state create; le tuple no. Se provi a cambiare un elemento di una tupla, Python genera un `TypeError`.

```python
# Con una lista SI possiamo riassegnare una posizione
A = ["disco", 10, 1.2]
print('Prima del cambio:', A)
A[0] = 'hard rock'
print('Dopo il cambio:', A)
```

```python
# Con una tupla, provare la stessa cosa genera un errore
tupla_prueba = ("disco", 10, 1.2)
try:
    tupla_prueba[0] = 'hard rock'
except TypeError as e:
    print("Errore! Le tuple non si possono modificare ->", e)
```

**Quando usare l'una o l'altra?** Regola pratica: se i dati cambieranno durante il programma, usa una **lista**. Se rappresentano qualcosa di fisso che è già accaduto o che non dovrebbe essere toccato, usa una **tupla**.

Una riga del dataset degli album (artista, anno, titolo) descrive un fatto storico — buona candidata a tupla. Un carrello della spesa che si riempie e si svuota — buona candidata a lista. Inoltre, essendo immutabili, le tuple sono leggermente più veloci e leggere in memoria.

## 2. Indicizzazione e slicing

Funziona allo stesso modo su liste e tuple.

```python
# Indicizzazione positiva: 0 è il primo elemento
L = ["Michael Jackson", 10.1, 1982]
print(L[0])
print(L[1])
print(L[2])
```

```python
# Indicizzazione negativa: -1 è l'ultimo elemento, contando dalla fine
print(L[-1])   # uguale a L[2]
print(L[-2])   # uguale a L[1]
print(L[-3])   # uguale a L[0]
```

```python
# Esattamente uguale con le tuple
tupla1 = ("disco", 10, 1.2)
print(tupla1[-1])
print(tupla1[-2])
```

**Slicing**: `sequenza[inizio:fine]` estrae un tratto. L'elemento nella posizione `fine` **non** è incluso.

```python
L = ["Michael Jackson", 10.1, 1982, "MJ", 1]
L[3:5]     # posizioni 3 e 4
```

```python
tupla2 = ("disco", 10, 1.2, "hard rock", 10)
print(tupla2[0:3])
print(tupla2[3:5])
print(len(tupla2))   # numero di elementi
```

Lo slicing di una lista restituisce una **lista nuova**; quello di una tupla restituisce una **tupla nuova**. Il tipo originale si conserva.

Inoltre, le tuple si possono **concatenare** con `+` (crea una tupla nuova, non modifica quelle originali) e si possono **ordinare** con `sorted()` (che, dato che la tupla non si può modificare, restituisce una **lista** nuova).

```python
# Concatenare tuple
tupla1 = ("disco", 10, 1.2)
tupla2 = tupla1 + ("hard rock", 10)
tupla2
```

```python
# sorted() su una tupla restituisce una lista nuova e ordinata
Ratings = (0, 9, 6, 5, 10, 8, 9, 6, 2)
RatingsSorted = sorted(Ratings)
print(RatingsSorted, type(RatingsSorted))
```

## 3. Metodi principali delle liste

| Metodo | Cosa fa |
|---|---|
| `append(x)` | Aggiunge **un solo** elemento in fondo |
| `extend(iterable)` | Aggiunge **ogni elemento** di un iterabile, separatamente |
| `insert(i, x)` | Inserisce `x` nella posizione `i` |
| `pop(i)` | Rimuove e restituisce l'elemento in `i` (senza argomento: l'ultimo) |
| `remove(x)` | Rimuove la prima occorrenza del **valore** `x` |
| `sort()` | Ordina la lista sul posto |
| `reverse()` | Inverte l'ordine sul posto |
| `del lista[i]` | Elimina l'elemento nella posizione `i` (istruzione, non metodo) |

`append` vs `extend` è la confusione più comune per chi inizia. `append` aggiunge l'argomento **come un unico blocco** (anche se è una lista). `extend` **appiattisce** l'iterabile e aggiunge ogni elemento singolarmente.

```python
# append: aggiunge UN elemento (che può essere una lista intera)
L = ["Michael Jackson", 10.2]
L.append(["pop", 10])
L
```

```python
# extend: aggiunge OGNI elemento separatamente (appiattisce)
L = ["Michael Jackson", 10.2]
L.extend(["pop", 10])
L
```

```python
# insert, pop, remove, sort, del in azione
L = ["Michael Jackson", 10.2, "pop", 10]
L.insert(2, "nuovo genere")
print("Dopo insert:", L)

elemento_quitado = L.pop()
print("Elemento rimosso con pop():", elemento_quitado)
print("Lista dopo pop():", L)

L.remove("nuovo genere")
print("Dopo remove('nuovo genere'):", L)

A = ["hard rock", 10, 1.2]
print('Prima di del:', A)
del A[0]
print('Dopo del:', A)
```

```python
# sort() ordina sul posto; sorted() restituisce una lista nuova
numeros = [5, 2, 9, 1]
numeros.sort()
print("sort() sul posto:", numeros)

letras = ["c", "a", "b"]
letras_ordenadas = sorted(letras)
print("sorted() -> nuova lista:", letras_ordenadas, "| originale intatta:", letras)
```

`split()` collega le stringhe alle liste: converte una stringa in una lista, separando per spazi (per default) o per il delimitatore che indichi. Molto utile quando si leggono dati grezzi, ad esempio una riga di un CSV.

```python
print('hard rock'.split())
print('Michael Jackson,Thriller,1982'.split(','))
```

## 4. Liste e tuple annidate (nesting)

Sia le liste che le tuple possono contenere, come qualsiasi altro elemento, un'altra lista o tupla. Ogni indice aggiuntivo ti fa scendere di un livello — come aprire sottocartelle dentro una cartella.

```python
# Una lista può mescolare tipi, includendo un'altra lista e una tupla
["Michael Jackson", 10.1, 1982, [1, 2], ("A", 1)]
```

```python
NestedT = (1, 2, ("pop", "rock"), (3, 4), ("disco", (1, 2)))

print("NestedT[2]:      ", NestedT[2])       # ('pop', 'rock')
print("NestedT[2][0]:   ", NestedT[2][0])    # 'pop'
print("NestedT[2][1]:   ", NestedT[2][1])    # 'rock'
print("NestedT[4]:      ", NestedT[4])       # ('disco', (1, 2))
print("NestedT[4][1]:   ", NestedT[4][1])    # (1, 2)
print("NestedT[4][1][0]:", NestedT[4][1][0]) # 1  -> tre livelli di profondità
print("NestedT[4][1][1]:", NestedT[4][1][1]) # 2
```

## 5. Copia per riferimento vs. clonazione — la trappola più comune

Quando scrivi `B = A` con le liste, **non** crei una seconda lista indipendente. Crei una seconda etichetta che punta allo **stesso oggetto** in memoria (copia per riferimento). Se modifichi `A`, il cambiamento si vede anche in `B`.

```python
A = ["hard rock", 10, 1.2]
B = A   # copia per riferimento: stessa lista, due nomi
print('A:', A)
print('B:', B)
```

```python
print('B[0] prima:', B[0])
A[0] = "banana"
print('B[0] dopo aver cambiato A:', B[0])   # è cambiato, anche se abbiamo toccato solo A!
```

Per ottenere una copia **vera**, indipendente, usiamo lo slicing completo `[:]` (o il metodo `.copy()`). Questo si chiama **clonare**.

```python
A = ["banana", 10, 1.2]
B = A[:]   # clonazione: una lista NUOVA con gli stessi valori
print('B[0] prima:', B[0])
A[0] = "hard rock"
print('B[0] dopo aver cambiato A:', B[0])   # B non è cambiata, è indipendente
```

Questo problema **non esiste con le tuple**: essendo immutabili, `B = A` non può mai fregarti, perché nessuna delle due etichette può comunque cambiare il contenuto. Un'altra ragione pratica per preferire le tuple quando i dati non hanno bisogno di cambiare.

## 6. Applicando tutto: il dataset degli album

Torniamo allo scenario iniziale: consigli di album da parte dei tuoi amici. Ogni album è una **tupla** (dato fisso, già accaduto), e la collezione di consigli è una **lista** (può crescere con nuovi consigli).

```python
# Ogni album: (Artista, Titolo, Anno, Genere, Valutazione degli amici)
album_thriller = ("Michael Jackson", "Thriller", 1982, "Pop, rock, R&B", 10.0)
album_back_in_black = ("AC/DC", "Back in Black", 1980, "Hard rock", 8.5)
album_dark_side = ("Pink Floyd", "The Dark Side of the Moon", 1973, "Progressive rock", 9.5)

# La lista dei consigli SI può crescere nel tempo
recomendaciones = [album_thriller, album_back_in_black]
print("Consigli iniziali:")
for album in recomendaciones:
    print(" -", album[1], "di", album[0], f"({album[2]})")

# Arriva un nuovo consiglio
recomendaciones.append(album_dark_side)
print("\nDopo il nuovo consiglio:")
for album in recomendaciones:
    print(" -", album[1], "di", album[0], f"({album[2]})")

# Provare a modificare un dato dell'album (una tupla) fallisce, come deve essere
try:
    album_thriller[2] = 1983
except TypeError as e:
    print("\nNon si può modificare un fatto storico:", e)
```

## Riepilogo della lezione

- **Sequenza ordinata**: collezione con posizioni fisse, accessibili per indice a partire da `0`. Liste e tuple sono le due di base.
- Le **liste** (`[ ]`) sono **mutabili**; le **tuple** (`( )`) sono **immutabili**.
- Usa le liste per dati che cambiano; le tuple per dati fissi (un fatto, una riga di registro storico).
- Indicizzazione (`L[0]`, `L[-1]`) e slicing (`L[inizio:fine]`) funzionano allo stesso modo su entrambe; lo slicing restituisce una sequenza nuova dello stesso tipo.
- `append()` aggiunge un solo blocco; `extend()` appiattisce e aggiunge ogni elemento. È la confusione più comune.
- `insert`, `pop`, `remove`, `sort`, `reverse` modificano la lista **sul posto**; `sorted()` restituisce una lista nuova senza toccare l'originale.
- Liste e tuple si possono **annidare**: ogni indice aggiuntivo scende di un livello (`L[2][0][1]`).
- `B = A` con le liste **non copia**, crea una seconda etichetta allo stesso oggetto. Per clonare davvero: `B = A[:]` oppure `A.copy()`.
- Questo problema di riferimento **non esiste nelle tuple**, perché non si possono comunque modificare.
