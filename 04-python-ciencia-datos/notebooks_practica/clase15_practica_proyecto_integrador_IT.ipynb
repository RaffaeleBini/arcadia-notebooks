```python
datos_poblacion = {
    'Pais': ['Cina', 'India', 'Brasile', 'Stati Uniti', 'Nigeria', 'Indonesia', 'Messico'],
    'Continente': ['Asia', 'Asia', 'America', 'America', 'Africa', 'Asia', 'America'],
    'PoblacionMillones': ['1412', '1408', '216', '335', '223', '277', '128'],
}

# Extract: dizionario -> DataFrame
df_poblacion = pd.DataFrame(datos_poblacion)

# Transform 1: la popolazione arriva come testo, la convertiamo a intero per poterci operare e ordinare
df_poblacion['PoblacionMillones'] = df_poblacion['PoblacionMillones'].astype(int)

# Transform 2: selezione booleana con .isin() -> teniamo solo Asia e America
#             (equivalente a: (df_poblacion['Continente'] == 'Asia') | (df_poblacion['Continente'] == 'America'))
df_poblacion = df_poblacion[df_poblacion['Continente'].isin(['Asia', 'America'])]

# Transform 3: ordiniamo dalla popolazione più alta alla più bassa
df_poblacion = df_poblacion.sort_values('PoblacionMillones', ascending=False)

# Transform 4: la somma totale si calcola sulla tabella GIÀ filtrata (solo Asia + America),
# proprio come nel progetto del PIL, dove prima ritagliamo e poi trasformiamo.
total_asia_america = df_poblacion['PoblacionMillones'].sum()
df_poblacion['PorcentajeDelTotal'] = np.round(
    df_poblacion['PoblacionMillones'] / total_asia_america * 100, 2
)

# Load: salviamo il risultato finale
df_poblacion.to_csv('reporte_poblacion.csv')

print(df_poblacion)
```
