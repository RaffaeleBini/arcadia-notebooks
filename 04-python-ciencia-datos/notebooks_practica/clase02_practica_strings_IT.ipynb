```python
titulo = "I Will Always Love You"
artista = "Whitney Houston"
album = "The BodyGuard"
anio_texto = "Lanzada en el año 1992 como parte del soundtrack"

# 1. f-string che combina le tre variabili
reporte = f"'{titulo}' di {artista}, dall'album {album}"

# 2. re.search cerca il pattern "quattro cifre di seguito" nel testo
import re
match_anio = re.search(r"\d\d\d\d", anio_texto)

# 3. Se c'e stato un match, match_anio.group() da il testo esatto trovato
if match_anio:
    anio_encontrado = match_anio.group()   # '1992'
else:
    anio_encontrado = None

# 4. f-string finale usando l'anno gia estratto
resumen_final = f"La canzone è del {anio_encontrado}"

print(reporte)
print("Anno trovato:", anio_encontrado)
print(resumen_final)
```
