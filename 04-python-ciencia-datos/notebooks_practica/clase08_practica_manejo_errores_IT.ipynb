```python
def safe_divide(numerador, denominador):
    try:
        # Se il denominatore è diverso da zero, questo si calcola senza problemi
        resultado = numerador / denominador
        return resultado
    except ZeroDivisionError:
        # Se il denominatore è 0, Python lancia ZeroDivisionError durante la divisione
        print("Errore: impossibile dividere per zero.")
        return None

# Casi di prova

print(safe_divide(-8, 4))   # -2.0
```

```python
def complex_calculation(num):
    try:
        # Se num == 5, questo lancia ZeroDivisionError.
        # Se num è una stringa come "abc", "abc" - 5 lancia TypeError.
        # except Exception cattura entrambi i casi allo stesso modo.
        result = num / (num - 5)
        print(f"Risultato: {result}")
    except Exception as e:
        print("Si è verificato un errore durante il calcolo.")

# Casi di prova
complex_calculation(10)     # Risultato: 2.0
complex_calculation(5)      # Si è verificato un errore durante il calcolo. (ZeroDivisionError)
complex_calculation("abc")  # Si è verificato un errore durante il calcolo. (TypeError)
```
