```python
import xml.etree.ElementTree as ET
import pandas as pd

empleados_datos = [
    {"nombre": "Lucia", "puesto": "Analista", "anios": 3},
    {"nombre": "Martin", "puesto": "Sviluppatore", "anios": 5},
    {"nombre": "Sofia", "puesto": "Data Scientist", "anios": 2},
]

raiz = ET.Element("empleados")

for empleado in empleados_datos:
    # TODO 1 risolto: sottoelemento "empleado" dentro la radice
    empleado_el = ET.SubElement(raiz, "empleado")

    # TODO 2 risolto: sottoelementi con il contenuto di ogni campo
    ET.SubElement(empleado_el, "nombre").text = empleado["nombre"]
    ET.SubElement(empleado_el, "puesto").text = empleado["puesto"]
    ET.SubElement(empleado_el, "anios").text = str(empleado["anios"])

arbol = ET.ElementTree(raiz)
arbol.write("empleados.xml")
print("File empleados.xml creato")

# TODO 3 risolto: rileggere l'XML in un DataFrame
df_empleados = pd.read_xml("empleados.xml", xpath="./empleado", parser="etree")
print(df_empleados)
```

**Suggerimenti:**
- `.info()` non ha bisogno di `print()`: basta scrivere `df.info()` in una cella.
- Per contare i valori mancanti per colonna puoi scorrere `df.isnull()` colonna per colonna con un `for`, oppure usare direttamente `df.isnull().sum()`.
- `.dtypes` è un attributo, non un metodo: si scrive senza parentesi.

````python
import numpy as np
import pandas as pd

np.random.seed(7)
n = 20

df_clinico = pd.DataFrame({
    "paciente_id": range(1, n + 1),
    "edad": np.random.randint(18, 80, size=n),
    "presion_sistolica": np.random.randint(90, 180, size=n).astype(float),
    "colesterol": np.random.normal(200, 30, size=n).round(1),
    "fumador": np.random.choice(["si", "no"], size=n),
})

# Valori mancanti inseriti apposta, per esercitarsi nel rilevamento
df_clinico.loc[[2, 5, 11], "colesterol"] = np.nan
df_clinico.loc[8, "presion_sistolica"] = np.nan

df_clinico.head()
````
