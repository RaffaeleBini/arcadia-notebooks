print(productos_tienda_b)

interseccion = productos_tienda_a.intersection(productos_tienda_b)
print("En común:", interseccion)   # {'Laptop', 'Auriculares'}

# Paso 4: diferencia -- productos exclusivos de la tienda A
solo_tienda_a = productos_tienda_a.difference(productos_tienda_b)
print("Solo en A:", solo_tienda_a)   # {'Teléfono', 'Tablet', 'Teclado'}

# Paso 5: unión -- catálogo combinado
catalogo_combinado = productos_tienda_a.union(productos_tienda_b)
print("Catálogo combinado:", catalogo_combinado)

# Paso 6: ¿{"Laptop", "Auriculares"} es subconjunto de productos_tienda_a?
es_subconjunto = {"Laptop", "Auriculares"}.issubset(productos_tienda_a)
print("¿Es subconjunto?:", es_subconjunto)   # True
