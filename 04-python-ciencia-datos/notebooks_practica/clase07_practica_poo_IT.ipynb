```python
class TextAnalyzer(object):

    def __init__(self, text):
        formattedText = text.replace('.', '').replace('!', '').replace('?', '').replace(',', '')
        formattedText = formattedText.lower()
        self.fmtText = formattedText

    def freqAll(self):
        wordList = self.fmtText.split(' ')
        freqMap = {}
        for word in set(wordList):
            freqMap[word] = wordList.count(word)
        return freqMap

    def freqOf(self, word):
        freqDict = self.freqAll()
        if word in freqDict:
            return freqDict[word]
        else:
            return 0

    def mostFrequent(self):
        freqDict = self.freqAll()
        # max() con key=... cerca la coppia (parola, frequenza) con la frequenza più alta
        parola, frequenza = max(freqDict.items(), key=lambda par: par[1])
        return (parola, frequenza)


recensione = "Il prodotto è molto affidabile. La spedizione è stata rapida e il prezzo è molto buono. Molto affidabile e molto rapido!"

analizzatore = TextAnalyzer(recensione)
print("Testo formattato:", analizzatore.fmtText)
print("Frequenza di 'affidabile':", analizzatore.freqOf("affidabile"))
print("Parola più frequente:", analizzatore.mostFrequent())
```
