precio_producto2 = 2300
precio_producto3 = 800

# Confirmamos el tipo de dato del resultado (debería ser int)
print(type(total_compra))

# Convertimos el número a texto para poder concatenarlo con un mensaje
total_compra_texto = str(total_compra)

# Armamos el mensaje final combinando texto con el valor convertido
print("El total de tu compra es " + total_compra_texto)
