# Passo 1 e 2: creare i set di ciascun negozio
productos_tienda_a = {"Telefono", "Laptop", "Tablet", "Auricolari", "Tastiera"}
productos_tienda_b = {"Laptop", "Monitor", "Auricolari", "Mouse"}

print(productos_tienda_a)
print(productos_tienda_b)

# Passo 3: intersezione -- prodotti venduti da entrambi i negozi
interseccion = productos_tienda_a.intersection(productos_tienda_b)
print("In comune:", interseccion)   # {'Laptop', 'Auricolari'}

# Passo 4: differenza -- prodotti esclusivi del negozio A
solo_tienda_a = productos_tienda_a.difference(productos_tienda_b)
print("Solo in A:", solo_tienda_a)   # {'Telefono', 'Tablet', 'Tastiera'}

# Passo 5: unione -- catalogo combinato
catalogo_combinado = productos_tienda_a.union(productos_tienda_b)
print("Catalogo combinato:", catalogo_combinado)

# Passo 6: {"Laptop", "Auricolari"} è sottoinsieme di productos_tienda_a?
es_subconjunto = {"Laptop", "Auricolari"}.issubset(productos_tienda_a)
print("È sottoinsieme?:", es_subconjunto)   # True
