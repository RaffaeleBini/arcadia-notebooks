<summary>💡 Vedi soluzione</summary>

```python
lista_compras = ["Orologio", "Portatile", "Scarpe", "Quaderno", "Palla"]

# --- Parte A: riferimento vs. clonazione ---

# 1. Riferimento: stessa lista con un altro nome. Clonazione: lista nuova e indipendente.
copia_referencia = lista_compras
copia_clonada = lista_compras[:]

# 2. Previsione (esempio di ragionamento atteso):
#    copia_referencia[0] -> "Bicicletta" (punta allo stesso oggetto di lista_compras)
#    copia_clonada[0]    -> "Orologio"   (è una copia indipendente, non si accorge del cambiamento)

# 3. Cambiamo l'originale
lista_compras[0] = "Bicicletta"

print("copia_referencia[0]:", copia_referencia[0])  # "Bicicletta" -> è cambiata insieme all'originale
print("copia_clonada[0]:", copia_clonada[0])         # "Orologio" -> non è cambiata, è indipendente

# --- Parte B: tupla di dati fissi ---

# 4. Tupla con i primi 3 articoli (già "decisi", non dovrebbero cambiare)
articulos_prioritarios = tuple(lista_compras[:3])
print("Articoli prioritari:", articulos_prioritarios)

# 5. Provare a modificare una tupla genera TypeError
try:
    articulos_prioritarios[0] = "Un'altra cosa"
except TypeError as e:
    print("Errore nel tentativo di modificare la tupla:", e)

# 6. Spiegazione:
# L'errore si verifica perché le tuple sono immutabili: una volta create, Python
# non permette di riassegnare nessuna delle loro posizioni. Ha senso usare una tupla
# qui perché "articulos_prioritarios" rappresenta una decisione già presa e fissa
# (come una riga di un registro storico); usare una tupla comunica questa intenzione
# ed evita che in un'altra parte del codice qualcuno la modifichi per errore, cosa
# che potrebbe invece succedere facilmente con una lista.
```
