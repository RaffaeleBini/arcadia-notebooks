<details>
<summary>💡 Vedi soluzione</summary>

```python
with open('poema.txt', 'r') as archivo:
    lineas = archivo.readlines()

total_lineas = len(lineas)

total_palabras = 0
for linea in lineas:
    palabras = linea.split()
    total_palabras += len(palabras)

print(f"Totale righe: {total_lineas}")
print(f"Totale parole: {total_palabras}")
```

<summary>💡 Vedi soluzione</summary>

```python
with open('socios.txt', 'r') as archivo:
    lineas = archivo.readlines()

encabezado = lineas[0]
filas = lineas[1:]

activos = []
bajas = []
for fila in filas:
    campos = fila.split()
    if campos[-1] == 'si':
        activos.append(fila)
    else:
        bajas.append(fila)

with open('activos.txt', 'w') as archivo:
    archivo.write(encabezado)
    for fila in activos:
        archivo.write(fila)

with open('bajas.txt', 'w') as archivo:
    archivo.write(encabezado)
    for fila in bajas:
        archivo.write(fila)
```

## Esercizio 3 — Filtrare un log di errori

Hai un file `log.txt` con righe di diversi livelli (`INFO`, `WARNING`, `ERROR`). Completa il codice per copiare solo le righe che contengono `"ERROR"` in un nuovo file `errores.txt`, usando il pattern di aprire due file contemporaneamente (uno per leggere, l'altro per scrivere).

**Suggerimenti:**
- Apri `log.txt` in modalità `'r'` e `errores.txt` in modalità `'w'`.
- Scorri il file di origine riga per riga con `for linea in archivo_log:`.
- Usa `if 'ERROR' in linea:` per decidere se quella riga va scritta nella destinazione.

````python
# Impalcatura: genera il file di input (non modificare questa parte)
with open('log.txt', 'w') as archivo:
    archivo.write("INFO: sistema avviato\n")
    archivo.write("ERROR: impossibile connettersi al database\n")
    archivo.write("INFO: nuovo tentativo di connessione\n")
    archivo.write("WARNING: memoria all'80%\n")
    archivo.write("ERROR: timeout scaduto\n")
    archivo.write("INFO: connessione stabilita\n")

# TODO: apri 'log.txt' in modalità lettura e 'errores.txt' in modalità scrittura (pattern copia/filtra)
# scorri 'log.txt' riga per riga e scrivi in 'errores.txt' solo le righe che contengono "ERROR"


# Verifica
with open('errores.txt', 'r') as archivo:
    print(archivo.read())
````

<details>
<summary>💡 Vedi soluzione</summary>

```python
with open('log.txt', 'r') as archivo_log:
    with open('errores.txt', 'w') as archivo_errores:
        for linea in archivo_log:
            if 'ERROR' in linea:
                archivo_errores.write(linea)

with open('errores.txt', 'r') as archivo:
    print(archivo.read())
```
