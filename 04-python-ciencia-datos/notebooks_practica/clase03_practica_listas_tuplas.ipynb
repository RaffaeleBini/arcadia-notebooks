```python
lista_compras = ["Reloj", "Notebook", "Zapatillas", "Cuaderno", "Pelota"]

# --- Parte A: referencia vs. clonado ---

# 1. Referencia: misma lista con otro nombre. Clonado: lista nueva e independiente.
copia_referencia = lista_compras
copia_clonada = lista_compras[:]

# 2. Predicción (ejemplo de razonamiento esperado):
#    copia_referencia[0] -> "Bicicleta" (apunta al mismo objeto que lista_compras)
#    copia_clonada[0]    -> "Reloj"     (es una copia independiente, no se entera del cambio)

# 3. Cambiamos el original
lista_compras[0] = "Bicicleta"

print("copia_referencia[0]:", copia_referencia[0])  # "Bicicleta" -> cambió junto con el original
print("copia_clonada[0]:", copia_clonada[0])         # "Reloj" -> no cambió, es independiente

# --- Parte B: tupla de datos fijos ---

# 4. Tupla con los primeros 3 artículos (ya "decididos", no deberían cambiar)
articulos_prioritarios = tuple(lista_compras[:3])
print("Artículos prioritarios:", articulos_prioritarios)

# 5. Intentar modificar una tupla lanza TypeError
try:
    articulos_prioritarios[0] = "Otra cosa"
except TypeError as e:
    print("Error al intentar modificar la tupla:", e)

# 6. Explicación:
# El error ocurre porque las tuplas son inmutables: una vez creadas, Python no
# permite reasignar ninguna de sus posiciones. Tiene sentido usar una tupla aquí
# porque "articulos_prioritarios" representa una decisión ya tomada y fija (como
# una fila de un registro histórico); usar una tupla comunica esa intención y
# evita que en otra parte del código alguien la modifique sin querer, algo que
# sí podría pasar fácilmente con una lista.
```
