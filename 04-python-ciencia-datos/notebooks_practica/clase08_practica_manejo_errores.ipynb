def safe_divide(numerador, denominador):
    try:
        # Si el denominador es distinto de cero, esto se calcula sin problema
        resultado = numerador / denominador
        return resultado
    except ZeroDivisionError:
        # Si el denominador es 0, Python lanza ZeroDivisionError al hacer la división
        print("Error: Cannot divide by zero.")
        return None

# Casos de prueba

def complex_calculation(num):
    try:
        # Si num == 5, esto lanza ZeroDivisionError.
        # Si num es un string como "abc", "abc" - 5 lanza TypeError.
        # except Exception atrapa cualquiera de los dos casos igual.
        result = num / (num - 5)
        print(f"Result: {result}")
    except Exception as e:
        print("An error occurred during calculation.")

# Casos de prueba
