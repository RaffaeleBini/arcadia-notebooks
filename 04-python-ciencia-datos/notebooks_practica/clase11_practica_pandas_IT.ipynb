## Esercizio 2 — `.loc` vs `.iloc`: lo stesso valore, due strade, e lo slicing che confonde

3. Verifica con `==` che entrambi i valori siano uguali.
4. Usa `.iloc` per selezionare l'intervallo di righe `1:4` (`df.iloc[1:4]`) e salva il risultato in `sub_iloc`.
5. Usa `.loc` per selezionare lo stesso intervallo di etichette `1:4` (`df.loc[1:4]`) e salva il risultato in `sub_loc`.
6. Stampa `len(sub_iloc)` e `len(sub_loc)`, e spiega (in un commento) perché non coincidono.

**Suggerimenti:**
- `.iloc[riga, colonna]` conta posizioni, come una lista di Python: il limite superiore resta escluso.
- `.loc[riga, colonna]` cerca etichette, come un dizionario: il limite superiore viene incluso.
- Con l'indice di default (0,1,2,3...), etichette e posizioni coincidono per una singola cella — ma non nello slicing.


````python
# TODO 1: valore nella riga di posizione 2, colonna di posizione 1, usando iloc
valor_iloc = None

# TODO 2: lo stesso valore, usando loc e il nome di colonna 'Album'
valor_loc = None

# TODO 3: confronta entrambi i valori
print(valor_iloc == valor_loc)

# TODO 4: seleziona l'intervallo di righe 1:4 usando iloc
sub_iloc = None

# TODO 5: seleziona l'intervallo di righe 1:4 usando loc
sub_loc = None

# TODO 6: quando finisci i TODO precedenti, decommenta la riga seguente
#         e stampa quante righe ha ciascuno, commentando perché differiscono
# print(len(sub_iloc), len(sub_loc))
````

<summary>💡 Vedi soluzione</summary>

```python
# 1. Valore nella riga di posizione 2, colonna di posizione 1 -> Album di Pink Floyd
valor_iloc = df.iloc[2, 1]

# 2. Lo stesso valore, usando .loc e il nome di colonna
valor_loc = df.loc[2, 'Album']

# 3. Devono essere uguali
print(valor_iloc == valor_loc)   # True

# 4. Con iloc, limite superiore ESCLUSO: posizioni 1, 2 e 3 (3 righe)
sub_iloc = df.iloc[1:4]

# 5. Con loc, limite superiore INCLUSO: etichette 1, 2, 3 e 4 (4 righe)
sub_loc = df.loc[1:4]

# 6. Numero di righe di ciascuno
print(len(sub_iloc), len(sub_loc))
# sub_iloc ha 3 righe perché iloc esclude la posizione 4.
# sub_loc ha 4 righe perché loc include l'etichetta 4 (il limite superiore).
```
