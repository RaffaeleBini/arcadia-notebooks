- Con el índice por defecto (0,1,2,3...), etiquetas y posiciones coinciden para una sola celda — pero no en el slicing.

```python
# 4. Con iloc, límite superior EXCLUIDO: posiciones 1, 2 y 3 (3 filas)

# 5. Con loc, límite superior INCLUIDO: etiquetas 1, 2, 3 y 4 (4 filas)
sub_loc = df.loc[1:4]

# 6. Cantidad de filas de cada uno
print(len(sub_iloc), len(sub_loc))
# sub_iloc tiene 3 filas porque iloc excluye la posición 4.
# sub_loc tiene 4 filas porque loc incluye la etiqueta 4 (el límite superior).
```
