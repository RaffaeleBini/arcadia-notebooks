<details>
<summary>💡 Ver solución</summary>

```python
with open('poema.txt', 'r') as archivo:
    lineas = archivo.readlines()

total_lineas = len(lineas)

total_palabras = 0
for linea in lineas:
    palabras = linea.split()
    total_palabras += len(palabras)

print(f"Total de líneas: {total_lineas}")
print(f"Total de palabras: {total_palabras}")
```

<summary>💡 Ver solución</summary>

```python
with open('socios.txt', 'r') as archivo:
    lineas = archivo.readlines()

encabezado = lineas[0]
filas = lineas[1:]

activos = []
bajas = []
for fila in filas:
    campos = fila.split()
    if campos[-1] == 'si':
        activos.append(fila)
    else:
        bajas.append(fila)

with open('activos.txt', 'w') as archivo:
    archivo.write(encabezado)
    for fila in activos:
        archivo.write(fila)

with open('bajas.txt', 'w') as archivo:
    archivo.write(encabezado)
    for fila in bajas:
        archivo.write(fila)

with open('activos.txt', 'r') as archivo:
    print("Socios activos:")
    print(archivo.read())

with open('bajas.txt', 'r') as archivo:
    print("Bajas:")
    print(archivo.read())
```

## Ejercicio 3 — Filtrar un log de errores

Tienes un archivo `log.txt` con líneas de distintos niveles (`INFO`, `WARNING`, `ERROR`). Completa el código para copiar solo las líneas que contienen `"ERROR"` hacia un nuevo archivo `errores.txt`, usando el patrón de abrir dos archivos a la vez (uno para leer, otro para escribir).

**Pistas:**
- Abre `log.txt` en modo `'r'` y `errores.txt` en modo `'w'`.
- Recorre el archivo de origen línea por línea con `for linea in archivo_log:`.
- Usa `if 'ERROR' in linea:` para decidir si esa línea se escribe en el destino.

````python
# Andamiaje: genera el archivo de entrada (no modifiques esta parte)
with open('log.txt', 'w') as archivo:
    archivo.write("INFO: sistema iniciado\n")
    archivo.write("ERROR: no se pudo conectar a la base de datos\n")
    archivo.write("INFO: reintentando conexión\n")
    archivo.write("WARNING: memoria al 80%\n")
    archivo.write("ERROR: tiempo de espera agotado\n")
    archivo.write("INFO: conexión establecida\n")

# TODO: abre 'log.txt' en modo lectura y 'errores.txt' en modo escritura (patrón de copiar/filtrar)
# recorre 'log.txt' línea por línea y escribe en 'errores.txt' solo las líneas que contienen "ERROR"


# Verificación
with open('errores.txt', 'r') as archivo:
    print(archivo.read())
````

<details>
<summary>💡 Ver solución</summary>

```python
with open('log.txt', 'r') as archivo_log:
    with open('errores.txt', 'w') as archivo_errores:
        for linea in archivo_log:
            if 'ERROR' in linea:
                archivo_errores.write(linea)

with open('errores.txt', 'r') as archivo:
    print(archivo.read())
```
