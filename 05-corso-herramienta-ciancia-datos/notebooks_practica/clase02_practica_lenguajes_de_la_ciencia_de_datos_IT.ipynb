```python
import pandas as pd

ventas = pd.DataFrame({
    "pais": ["España", "México", "España", "Argentina"],
    "unidades": [80, 40, 30, 90],
})

resultado = ventas[(ventas["pais"] == "España") & (ventas["unidades"] > 50)]
```
