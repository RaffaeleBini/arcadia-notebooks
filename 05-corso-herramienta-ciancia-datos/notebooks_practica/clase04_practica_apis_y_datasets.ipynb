import requests

url = "https://api.open-meteo.com/v1/forecast"
parametros = {
    "latitude": 40.42,   # ejemplo: Madrid
    "longitude": -3.70,
    "current_weather": "true",
}

respuesta = requests.get(url, params=parametros)
datos = respuesta.json()

print("Temperatura:", datos["current_weather"]["temperature"], "°C")
print("Velocidad del viento:", datos["current_weather"]["windspeed"], "km/h")
