<summary>💡 Ver solución</summary>

```python
import pandas as pd

ventas = pd.DataFrame({
    "pais": ["España", "México", "España", "Argentina"],
    "unidades": [80, 40, 30, 90],
})

resultado = ventas[(ventas["pais"] == "España") & (ventas["unidades"] > 50)]

# Equivalente en SQL:
# SELECT * FROM ventas WHERE pais = 'España' AND unidades > 50;

print(resultado)
```
