## 1. Crear cuenta y repositorio (demostración en github.com)

1. Ir a github.com/join: email, contraseña, usuario, verificación.
2. Clic en "+" -> "New repository": nombre, público/privado, inicializar con README.
3. Pestañas del repositorio: Code, Issues, Pull Requests, Projects, Wiki, Settings.

## 2. Editar, crear y subir archivos (demostración en github.com)

- Editar: icono de lápiz sobre un archivo -> modificar -> "Commit changes".
- Crear: "Add file" -> "Create new file" -> nombre + contenido -> "Commit changes".
- Subir: "Add file" -> "Upload files" -> arrastrar archivos -> "Commit changes".

## 3. Ramas y Pull Requests: el mismo flujo, en Git real

Todo lo que harías con clics en GitHub (crear rama, comprometer cambios, comparar,
abrir y fusionar un Pull Request) tiene un equivalente exacto en línea de comandos,
que sí podemos ejecutar aquí:

```python
!mkdir -p /content/repo_github
%cd /content/repo_github
!git init -q
!git config user.email "estudiante@ejemplo.com"
!git config user.name "Estudiante del curso"

!echo "# Mi proyecto" > README.md
!git add README.md
!git commit -q -m "Commit inicial: añade README"
```

```python
# 1. Crear una rama (equivalente a "Branch: main" -> escribir nombre -> Create branch)
!git checkout -q -b rama_nueva_funcionalidad

# 2. Confirmar cambios en la rama (equivalente a "Add file > Create new file" en la rama)
!echo "print('Nueva funcionalidad')" > nueva_funcionalidad.py
!git add nueva_funcionalidad.py
!git commit -q -m "Añade nueva_funcionalidad.py"
```

```python
# 3. Comparar con main (equivalente a "Compare & pull request")
!git diff main rama_nueva_funcionalidad --stat
```

```python
# 4. Fusionar (equivalente a "Merge pull request" + "Confirm merge")
!git checkout -q main
!git merge -q rama_nueva_funcionalidad -m "Merge pull request: nueva funcionalidad"
!ls
```
