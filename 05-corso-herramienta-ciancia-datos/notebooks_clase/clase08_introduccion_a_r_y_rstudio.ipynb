## 1. RStudio, el IDE de R

RStudio unifica editor, consola, historial de variables (pestañas Environment/History)
y gestión de paquetes (Files/Plots/Packages/Help) en una sola ventana — el mismo
papel que Jupyter cumple para Python.

Paquetes de R muy usados: `dplyr` (manipulación de datos, equivalente a pandas),
`stringr` (texto), `ggplot2` (visualización, clase 9) y `caret` (Machine Learning).

## 2. El código original en R

Así es como se ve, en R, la carga y primera exploración del dataset `iris`
(incluido en R por defecto) — tal como aparece en el material fuente de instalación
de R:

```r
library(datasets)
data(iris)
View(iris)
unique(iris$Species)
```

## 3. El mismo flujo, en Python

`library()` en R equivale a `import` en Python; `iris$Species` (acceso a columna con
`$`) equivale a `iris["Species"]` en pandas.

````python
from sklearn.datasets import load_iris
import pandas as pd

# Cargamos el mismo dataset iris, esta vez a través de scikit-learn
iris_sklearn = load_iris()
iris = pd.DataFrame(iris_sklearn.data, columns=iris_sklearn.feature_names)
iris["Species"] = pd.Categorical.from_codes(iris_sklearn.target, iris_sklearn.target_names)

# Equivalente a View(iris) en RStudio: mostramos las primeras filas
iris.head()
````

````python
# Equivalente a unique(iris$Species) en R
especies = iris["Species"].unique()
print(especies)
````

## Resumen

- RStudio es al lenguaje R lo que Jupyter es a Python: el entorno de trabajo diario.
- dplyr, stringr, ggplot2 y caret son los equivalentes en R a pandas, manejo de
  texto, matplotlib/seaborn y scikit-learn.
- El mismo dataset `iris` sirve de puente entre ambos lenguajes: `library()` ~
  `import`, `iris$columna` ~ `iris["columna"]`.
