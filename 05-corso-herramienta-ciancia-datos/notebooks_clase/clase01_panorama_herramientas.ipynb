## 1. Las seis tareas de la Ciencia de Datos

Todo proyecto de datos pasa, en algún momento, por estas seis tareas:

1. **Gestión de datos** — recolectar, guardar y recuperar datos.
2. **Integración y transformación (ETL/ELT)** — limpiar y unificar datos de varias fuentes.
3. **Visualización** — convertir tablas en gráficos que se entienden de un vistazo.
4. **Construcción de modelos** — entrenar algoritmos que aprenden patrones.
5. **Despliegue de modelos** — poner el modelo a disposición de otras aplicaciones (vía API).
6. **Monitoreo y evaluación** — vigilar que el modelo siga siendo preciso y justo.

Estas seis tareas se apoyan en cuatro capas: gestión de activos de código, gestión de
activos de datos, entornos de ejecución y entornos de desarrollo (IDE).

```python
import pandas as pd

# 1. Gestión de datos
datos = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")

# 2. Integración y transformación (ETL)
datos = datos.rename(columns={"total_bill": "cuenta_total"})

# 3. Visualización
datos.groupby("day")["cuenta_total"].sum().plot(kind="bar", title="Cuenta total por día")
```

## 2. El mismo mapa, tres estanterías

Las mismas seis tareas se resuelven con herramientas **open source**, **comerciales** o
**en la nube**. Algunos ejemplos que ya viste en el manual (capítulo 3):

| Categoría | Open source | Comercial | Nube |
|---|---|---|---|
| Gestión de datos | MySQL, MongoDB | Oracle DB, Db2 | Amazon DynamoDB, IBM Cloudant |
| ETL | Apache Airflow, Kafka | Informatica, Talend | Informatica Cloud, IBM Data Refinery |
| Visualización | Apache Superset, Kibana | Tableau, Power BI | IBM Cognos, Datameer |
| Modelos | scikit-learn, TensorFlow | SPSS Modeler, SAS | Watson ML, Azure ML |
| Despliegue | TensorFlow Serving, Seldon | SPSS Collaboration | API REST en la nube |
| Monitoreo | Prometheus, AI Fairness 360 | — | Watson OpenScale, SageMaker Model Monitor |
| Código | Git, GitHub, GitLab | — | — |

```python
# Un pequeño catálogo de herramientas para explorar en vivo durante la clase
catalogo = {
    "MySQL": ("Gestión de datos", "open source"),
    "MongoDB": ("Gestión de datos", "open source"),
    "Apache Airflow": ("Integración/ETL", "open source"),
    "Tableau": ("Visualización", "comercial"),
    "scikit-learn": ("Modelos", "open source"),
    "Watson OpenScale": ("Monitoreo", "nube"),
    "GitHub": ("Gestión de código", "nube/comercial"),
}

for herramienta, (categoria, tipo) in catalogo.items():
    print(f"{herramienta:20s} -> {categoria:20s} ({tipo})")
```
