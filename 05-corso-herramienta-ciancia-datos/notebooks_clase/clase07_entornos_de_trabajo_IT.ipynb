## 1. Kernel

Un kernel è il motore che esegue il tuo codice. Puoi cambiarlo dal menu del
notebook (Python, R, Julia...). Riavviarlo (`Runtime > Restart runtime` su Colab)
risolve la maggior parte dei comportamenti strani, perché cancella tutte le variabili in
memoria e ricomincia da zero.

```python
import sys
print("Ejecutable de Python en uso:", sys.executable)
print("Versión:", sys.version)
```

## 2. Anaconda, JupyterLab e VS Code (ripasso concettuale)

- **Anaconda**: distribuzione di Python/R con più di 1.500 librerie preinstallate e
  Anaconda Navigator come interfaccia grafica.
- **JupyterLab**: interfaccia più completa del Jupyter classico, con più file
  aperti contemporaneamente.
- **VS Code**: editor generale di Microsoft a cui si aggiunge supporto per i notebook
  tramite estensioni.

## 3. JupyterLite vs. Google Colab

- **JupyterLite**: gira interamente nel browser (Pyodide/WebAssembly), senza
  server. Ideale per esercizi leggeri.
- **Google Colab**: esegue il tuo codice su un server reale di Google, con GPU
  disponibile, integrazione con Drive e GitHub.

```python
# Rileviamo in quale ambiente si sta eseguendo questo notebook
try:
    import google.colab
    entorno = "Google Colab"
except ImportError:
    entorno = "Jupyter local (o JupyterLite)"

print(f"Este notebook se está ejecutando en: {entorno}")
```

```python
# Installare una libreria aggiuntiva funziona allo stesso modo in Colab e in Jupyter locale
# (si esegue senza errori anche se la libreria è già installata)
!pip install --quiet requests
import requests
print("Librería 'requests' lista para usar, versión:", requests.__version__)
```

## Riepilogo

- Il kernel esegue il tuo codice; riavviarlo è la soluzione più comune a comportamenti
  strani.
- Anaconda/VS Code sono per il lavoro locale; JupyterLite e Colab sono per lavorare senza
  installare nulla.
- `!pip install` funziona allo stesso modo in Colab e in Jupyter locale per aggiungere librerie.
