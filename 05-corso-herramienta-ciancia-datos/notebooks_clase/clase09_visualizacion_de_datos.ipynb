# Clase 9 — Visualización de datos (de ggplot2 a Python)

**Objetivos de la clase:**
- Identificar la lógica de "gramática de gráficos" de ggplot2.
- Construir el equivalente de un gráfico de dispersión con título y etiquetas usando
  Matplotlib/Seaborn.

Basado en el capítulo 17 del manual.

**Nota:** el código ggplot2 se muestra en Markdown, fiel al material fuente. Las
celdas ejecutables reproducen el mismo gráfico y dataset (`mtcars`) con
Matplotlib/Seaborn en Python.

## 1. El código original en R (ggplot2)

```r
install.packages("ggplot2")
library(ggplot2)

ggplot(mtcars, aes(x = wt, y = mpg)) +
  geom_point() +
  ggtitle("Relación entre peso y consumo de combustible") +
  labs(x = "Peso del coche (miles de libras)", y = "Millas por galón")
```

ggplot2 construye el gráfico por capas: datos y ejes (`aes`), tipo de gráfico
(`geom_point`), y detalles estéticos (`ggtitle`, `labs`).

## 2. El mismo gráfico, en Python

El dataset `mtcars` no viene con Python, pero es un CSV público muy usado en
tutoriales; lo cargamos directamente desde su URL.

````python
import pandas as pd
import matplotlib.pyplot as plt

url_mtcars = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
mtcars = pd.read_csv(url_mtcars)

# En este dataset equivalente: "weight" (peso) y "mpg" (millas por galón)
plt.scatter(mtcars["weight"], mtcars["mpg"])
plt.title("Relación entre peso y consumo de combustible")
plt.xlabel("Peso del coche (lb)")
plt.ylabel("Millas por galón")
plt.show()
````

## 3. La versión con Seaborn

Seaborn añade, con menos código, elementos estadísticos como una línea de tendencia.

````python
import seaborn as sns

sns.regplot(data=mtcars, x="weight", y="mpg")
plt.title("Peso vs. consumo, con línea de tendencia")
plt.show()
````

## Resumen

- ggplot2 construye un gráfico por capas (`aes` + `geom_...` + `ggtitle`/`labs`).
- Matplotlib reproduce el mismo resultado con `plt.scatter()`, `plt.title()`,
  `plt.xlabel()`, `plt.ylabel()`.
- Seaborn añade capas estadísticas (como una línea de tendencia) con menos código
  que Matplotlib puro.
