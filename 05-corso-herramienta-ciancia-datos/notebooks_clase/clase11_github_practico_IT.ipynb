## 1. Creare account e repository (dimostrazione su github.com)

1. Andare su github.com/join: email, password, nome utente, verifica.
2. Clic su "+" -> "New repository": nome, pubblico/privato, inizializzare con README.
3. Schede del repository: Code, Issues, Pull Requests, Projects, Wiki, Settings.

## 2. Modificare, creare e caricare file (dimostrazione su github.com)

- Modificare: icona a matita su un file -> modificare -> "Commit changes".
- Creare: "Add file" -> "Create new file" -> nome + contenuto -> "Commit changes".
- Caricare: "Add file" -> "Upload files" -> trascinare file -> "Commit changes".

## 3. Branch e Pull Request: lo stesso flusso, in Git reale

Tutto ciò che faresti con clic su GitHub (creare branch, confermare modifiche, confrontare,
aprire e fondere una Pull Request) ha un equivalente esatto a riga di comando,
che possiamo davvero eseguire qui:

```python
!mkdir -p /content/repo_github
%cd /content/repo_github
!git init -q
!git config user.email "estudiante@ejemplo.com"
!git config user.name "Estudiante del curso"

!echo "# Mi proyecto" > README.md
!git add README.md
!git commit -q -m "Commit inicial: añade README"
```

```python
# 1. Creare un branch (equivalente a "Branch: main" -> scrivere nome -> Create branch)
!git checkout -q -b rama_nueva_funcionalidad

# 2. Confermare modifiche nel branch (equivalente a "Add file > Create new file" nel branch)
!echo "print('Nueva funcionalidad')" > nueva_funcionalidad.py
!git add nueva_funcionalidad.py
!git commit -q -m "Añade nueva_funcionalidad.py"
```

```python
# 3. Confrontare con main (equivalente a "Compare & pull request")
!git diff main rama_nueva_funcionalidad --stat
```

```python
# 4. Fondere (equivalente a "Merge pull request" + "Confirm merge")
!git checkout -q main
!git merge -q rama_nueva_funcionalidad -m "Merge pull request: nueva funcionalidad"
!ls
```
