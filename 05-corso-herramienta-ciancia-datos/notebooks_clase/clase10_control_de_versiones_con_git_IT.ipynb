## 1. Perché esiste il controllo di versione

Git registra tutte le modifiche di un progetto per poter consultare la cronologia,
confrontare versioni o tornare indietro se qualcosa va storto. È un sistema **distribuito**:
ogni copia ha la cronologia completa, non dipende da un unico server centrale.

Comandi essenziali: `git init/clone`, `git status`, `git add`, `git commit`,
`git log`, `git branch`, `git checkout`, `git merge`, `git reset`.

```python
# Google Colab permette di eseguire comandi di terminale anteponendo "!"
!git --version
```

## 2. Il ciclo base: init, add, commit, log

```python
!mkdir -p /content/mi_proyecto
%cd /content/mi_proyecto
!git init
!git config user.email "estudiante@ejemplo.com"
!git config user.name "Estudiante del curso"
```

```python
# Creiamo un file e verifichiamo cosa rileva Git
!echo "print('hola mundo')" > script.py
!git status
```

```python
# Confermiamo la modifica nella cronologia (commit)
!git add script.py
!git commit -m "Primer commit: añade script.py"
!git log --oneline
```

## Riepilogo

- Git è un sistema di controllo di versione distribuito.
- Il ciclo base è: `git init` (o `clone`) -> `git add` -> `git commit` -> `git log`.
- `git status` ti dice in ogni momento cosa è cambiato dall'ultimo commit.
- Puoi esercitarti con Git reale dentro un notebook anteponendo `!` a ogni comando.
