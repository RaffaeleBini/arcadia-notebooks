## 3. Commenti in Python

Tutto ciò che segue `#` in una riga è un commento: non si esegue, è solo un appunto
per chi legge il codice.

```python
# Questa riga è un commento, non fa nulla eseguendo la cella
resultado = 3 + 3  # puoi anche commentare alla fine di una riga
print(resultado)
```

## 4. Gestire il file del notebook

- **Eseguire tutto**: `Runtime > Run all` (Colab) o `Cell > Run All` (Jupyter classico).
- **Rinominare**: clic sul nome del notebook nella parte superiore.
- **Riavviare il kernel**: se qualcosa si comporta in modo strano, `Runtime > Restart runtime`
  (Colab) è l'equivalente di "spegnere e riaccendere".
- **Salvare e scaricare**: `File > Download > Download .ipynb`.

Una buona pratica è lasciare, come prima cella di qualsiasi nuovo notebook, una
cella di "configurazione":
