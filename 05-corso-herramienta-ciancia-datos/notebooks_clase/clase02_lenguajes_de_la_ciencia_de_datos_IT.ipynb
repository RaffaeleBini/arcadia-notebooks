## 2. R e SQL: due approcci diversi

- **R** è il linguaggio di riferimento in statistica, con più di 15.000 pacchetti.
  Si distribuisce sotto licenza GPL (software libero secondo la FSF, open source secondo l'OSI).
- **SQL** è un linguaggio **non procedurale**: descrivi *quale* risultato vuoi, non
  *come* ottenerlo. Lavora su database relazionali (tabelle con righe e
  colonne) ed è standardizzato dall'ANSI.

Anche se in questo corso lavoriamo in Python, ecco lo stesso filtro espresso nei
tre stili, solo per confrontare:

```python
import pandas as pd

datos = pd.DataFrame({
    "producto": ["A", "B", "C", "D"],
    "precio": [50, 120, 30, 200],
})

# Stile Python/pandas: istruzioni passo per passo ("come")
caros_python = datos[datos["precio"] > 100]
print(caros_python)

# Stile SQL equivalente (solo come riferimento, non si esegue qui):
# SELECT * FROM datos WHERE precio > 100;

# Stile R equivalente (solo come riferimento, non si esegue qui):
# datos[datos$precio > 100, ]
```

## 3. Altri linguaggi dell'ecosistema

- **Java**: gira sulla JVM; appare in Weka, Java-ML, Apache MLlib, Hadoop.
- **Scala**: linguaggio nativo di Apache Spark (Shark, MLlib, GraphX, Spark Streaming).
- **C++**: usato internamente in librerie veloci come TensorFlow, MongoDB, Caffe.
- **JavaScript**: TensorFlow.js, brain.js — Machine Learning nel browser.
- **Julia**: calcolo numerico ad alte prestazioni (JuliaDB).

## Riepilogo

- Python vince per le sue librerie, non per design originale.
- SQL descrive *cosa* vuoi (non procedurale); Python descrive *come* ottenerlo.
- Ogni linguaggio ha la sua nicchia: Java/Scala in Hadoop/Spark, C++ nelle prestazioni,
  JavaScript nel browser, Julia nel calcolo numerico.
