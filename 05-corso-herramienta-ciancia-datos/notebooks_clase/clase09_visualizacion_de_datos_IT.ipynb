- Costruire l'equivalente di un grafico a dispersione con titolo ed etichette usando

Basato sul capitolo 17 del manuale.

**Nota:** il codice ggplot2 si mostra in Markdown, fedele al materiale sorgente. Le
celle eseguibili riproducono lo stesso grafico e dataset (`mtcars`) con
Matplotlib/Seaborn in Python.

## 1. Il codice originale in R (ggplot2)

```r
install.packages("ggplot2")
library(ggplot2)

ggplot(mtcars, aes(x = wt, y = mpg)) +
  geom_point() +
  ggtitle("Relación entre peso y consumo de combustible") +
  labs(x = "Peso del coche (miles de libras)", y = "Millas por galón")
```

ggplot2 costruisce il grafico a strati: dati e assi (`aes`), tipo di grafico
(`geom_point`), e dettagli estetici (`ggtitle`, `labs`).

## 2. Lo stesso grafico, in Python

Il dataset `mtcars` non arriva con Python, ma è un CSV pubblico molto usato nei
tutorial; lo carichiamo direttamente dal suo URL.

````python
import pandas as pd
import matplotlib.pyplot as plt

url_mtcars = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
mtcars = pd.read_csv(url_mtcars)

# In questo dataset equivalente: "weight" (peso) e "mpg" (miglia per gallone)
plt.scatter(mtcars["weight"], mtcars["mpg"])
plt.title("Relación entre peso y consumo de combustible")
plt.xlabel("Peso del coche (lb)")
plt.ylabel("Millas por galón")
plt.show()
````

## 3. La versione con Seaborn

Seaborn aggiunge, con meno codice, elementi statistici come una linea di tendenza.

````python
import seaborn as sns

sns.regplot(data=mtcars, x="weight", y="mpg")
plt.title("Peso vs. consumo, con línea de tendencia")
plt.show()
````

## Riepilogo

- ggplot2 costruisce un grafico a strati (`aes` + `geom_...` + `ggtitle`/`labs`).
- Matplotlib riproduce lo stesso risultato con `plt.scatter()`, `plt.title()`,
  `plt.xlabel()`, `plt.ylabel()`.
- Seaborn aggiunge strati statistici (come una linea di tendenza) con meno codice
  di Matplotlib puro.
