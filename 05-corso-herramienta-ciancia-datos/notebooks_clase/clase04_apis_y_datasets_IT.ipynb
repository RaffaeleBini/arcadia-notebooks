## 1. Il pattern di un'API

Client -> Richiesta (Request) -> Endpoint -> Risposta (Response, normalmente in JSON).

Rendiamolo reale con un'API pubblica gratuita che non richiede chiave di accesso:
Open-Meteo, che restituisce il meteo attuale di qualsiasi coordinata.

```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
parametros = {
    "latitude": 40.71,
    "longitude": -74.01,
    "current_weather": "true",
}

respuesta = requests.get(url, params=parametros)
datos = respuesta.json()

print("Código de estado HTTP:", respuesta.status_code)
print("Temperatura actual en Nueva York:", datos["current_weather"]["temperature"], "°C")
```

## 2. Dataset e licenze

Un dataset può essere tabulare (CSV), gerarchico o di rete/grafo. Esistono cataloghi
enormi di dati aperti: data.gov, kaggle.com/datasets, data.worldbank.org, who.int.

Ogni dataset ha una licenza (Dominio Pubblico, CC-BY, CDLA-Sharing, CDLA-Permissive...)
che definisce cosa puoi farci. Il dataset meteorologico che useremo nel
capitolo di chiusura del corso è sotto licenza CDLA-Sharing.

```python
import pandas as pd

url_clima = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/fJKLoMKtgpDAl5MkA6rf2A/jfk-weather.csv"
clima = pd.read_csv(url_clima, parse_dates=["DATE"], low_memory=False)

print(f"Filas: {clima.shape[0]}, columnas: {clima.shape[1]}")
clima.head()
```
