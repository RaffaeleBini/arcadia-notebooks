## 1. Kernels

Un kernel es el motor que ejecuta tu código. Puedes cambiarlo desde el menú del
notebook (Python, R, Julia...). Reiniciarlo (`Runtime > Restart runtime` en Colab)
resuelve la mayoría de comportamientos extraños, porque borra todas las variables en
memoria y empieza de cero.

```python
import sys
print("Ejecutable de Python en uso:", sys.executable)
print("Versión:", sys.version)
```

## 2. Anaconda, JupyterLab y VS Code (repaso conceptual)

- **Anaconda**: distribución de Python/R con más de 1.500 librerías preinstaladas y
  Anaconda Navigator como interfaz gráfica.
- **JupyterLab**: interfaz más completa que el Jupyter clásico, con varios archivos
  abiertos a la vez.
- **VS Code**: editor general de Microsoft al que se le añade soporte para notebooks
  mediante extensiones.

## 3. JupyterLite vs. Google Colab

- **JupyterLite**: corre enteramente en el navegador (Pyodide/WebAssembly), sin
  servidor. Ideal para ejercicios ligeros.
- **Google Colab**: ejecuta tu código en un servidor real de Google, con GPU
  disponible, integración con Drive y GitHub.

```python
# Detectamos en qué entorno se está ejecutando este notebook
try:
    import google.colab
    entorno = "Google Colab"
except ImportError:
    entorno = "Jupyter local (o JupyterLite)"

print(f"Este notebook se está ejecutando en: {entorno}")
```

```python
# Instalar una librería adicional funciona igual en Colab que en Jupyter local
# (se ejecuta sin error incluso si la librería ya está instalada)
!pip install --quiet requests
import requests
print("Librería 'requests' lista para usar, versión:", requests.__version__)
```

## Resumen

- El kernel ejecuta tu código; reiniciarlo es la solución más común a comportamientos
  extraños.
- Anaconda/VS Code son para trabajo local; JupyterLite y Colab son para trabajar sin
  instalar nada.
- `!pip install` funciona igual en Colab que en Jupyter local para añadir librerías.
