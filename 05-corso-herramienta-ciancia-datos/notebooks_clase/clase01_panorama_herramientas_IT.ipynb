## 1. Le sei attività della Scienza dei Dati

Ogni progetto di dati attraversa, in qualche momento, queste sei attività:

1. **Gestione dei dati** — raccogliere, salvare e recuperare dati.
2. **Integrazione e trasformazione (ETL/ELT)** — pulire e unificare dati da diverse fonti.
3. **Visualizzazione** — trasformare tabelle in grafici comprensibili a colpo d'occhio.
4. **Costruzione di modelli** — addestrare algoritmi che imparano pattern.
5. **Distribuzione di modelli** — mettere il modello a disposizione di altre applicazioni (via API).
6. **Monitoraggio e valutazione** — vigilare che il modello resti preciso ed equo.

Queste sei attività si appoggiano su quattro livelli: gestione degli asset di codice, gestione
degli asset di dati, ambienti di esecuzione e ambienti di sviluppo (IDE).

```python
import pandas as pd

# 1. Gestione dei dati
datos = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")

# 2. Integrazione e trasformazione (ETL)
datos = datos.rename(columns={"total_bill": "cuenta_total"})

# 3. Visualizzazione
datos.groupby("day")["cuenta_total"].sum().plot(kind="bar", title="Cuenta total por día")
```
