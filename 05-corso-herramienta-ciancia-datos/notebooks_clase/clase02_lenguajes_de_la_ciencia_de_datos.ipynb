## 2. R y SQL: dos enfoques distintos

- **R** es el lenguaje de referencia en estadística, con más de 15.000 paquetes.
  Se distribuye bajo licencia GPL (software libre según la FSF, open source según la OSI).
- **SQL** es un lenguaje **no procedural**: describes *qué* resultado quieres, no
  *cómo* conseguirlo. Trabaja sobre bases de datos relacionales (tablas con filas y
  columnas) y está estandarizado por ANSI.

Aunque en este curso trabajamos en Python, aquí tienes el mismo filtro expresado en
los tres estilos, solo para comparar:

```python
import pandas as pd

datos = pd.DataFrame({
    "producto": ["A", "B", "C", "D"],
    "precio": [50, 120, 30, 200],
})

# Estilo Python/pandas: instrucciones paso a paso ("cómo")
caros_python = datos[datos["precio"] > 100]
print(caros_python)

# Estilo SQL equivalente (solo como referencia, no se ejecuta aquí):
# SELECT * FROM datos WHERE precio > 100;

# Estilo R equivalente (solo como referencia, no se ejecuta aquí):
# datos[datos$precio > 100, ]
```

## 3. Otros lenguajes del ecosistema

- **Java**: corre sobre la JVM; aparece en Weka, Java-ML, Apache MLlib, Hadoop.
- **Scala**: lenguaje nativo de Apache Spark (Shark, MLlib, GraphX, Spark Streaming).
- **C++**: usado por dentro en librerías rápidas como TensorFlow, MongoDB, Caffe.
- **JavaScript**: TensorFlow.js, brain.js — Machine Learning en el navegador.
- **Julia**: cálculo numérico de alto rendimiento (JuliaDB).

## Resumen

- Python gana por sus librerías, no por diseño original.
- SQL describe *qué* quieres (no procedural); Python describe *cómo* conseguirlo.
- Cada lenguaje tiene su nicho: Java/Scala en Hadoop/Spark, C++ en rendimiento,
  JavaScript en el navegador, Julia en cálculo numérico.
