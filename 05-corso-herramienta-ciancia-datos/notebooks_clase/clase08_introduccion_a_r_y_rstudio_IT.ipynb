## 1. RStudio, l'IDE di R

RStudio unifica editor, console, cronologia delle variabili (schede Environment/History)
e gestione dei pacchetti (Files/Plots/Packages/Help) in un'unica finestra — lo stesso
ruolo che Jupyter svolge per Python.

Pacchetti R molto usati: `dplyr` (manipolazione dei dati, equivalente a pandas),
`stringr` (testo), `ggplot2` (visualizzazione, lezione 9) e `caret` (Machine Learning).

## 2. Il codice originale in R

Ecco come appare, in R, il caricamento e la prima esplorazione del dataset `iris`
(incluso in R di default) — così come appare nel materiale sorgente di installazione
di R:

```r
library(datasets)
data(iris)
View(iris)
unique(iris$Species)
```

## 3. Lo stesso flusso, in Python

`library()` in R equivale a `import` in Python; `iris$Species` (accesso a colonna con
`$`) equivale a `iris["Species"]` in pandas.

````python
from sklearn.datasets import load_iris
import pandas as pd

# Carichiamo lo stesso dataset iris, questa volta tramite scikit-learn
iris_sklearn = load_iris()
iris = pd.DataFrame(iris_sklearn.data, columns=iris_sklearn.feature_names)
iris["Species"] = pd.Categorical.from_codes(iris_sklearn.target, iris_sklearn.target_names)

# Equivalente a View(iris) in RStudio: mostriamo le prime righe
iris.head()
````

````python
# Equivalente a unique(iris$Species) in R
especies = iris["Species"].unique()
print(especies)
````

## Riepilogo

- RStudio è al linguaggio R ciò che Jupyter è a Python: l'ambiente di lavoro quotidiano.
- dplyr, stringr, ggplot2 e caret sono gli equivalenti in R di pandas, gestione del
  testo, matplotlib/seaborn e scikit-learn.
- Lo stesso dataset `iris` fa da ponte tra i due linguaggi: `library()` ~
  `import`, `iris$colonna` ~ `iris["colonna"]`.
