## 1. Celdas de código y celdas de Markdown

Un notebook combina dos tipos de celda:

- **Código**: se ejecuta con `Shift + Enter`; el resultado aparece justo debajo.
- **Markdown**: texto con formato (como esta misma celda), para explicar el código.

Prueba a ejecutar la siguiente celda de código con `Shift + Enter`:

```python
# Esto es una celda de código. Ejecútala con Shift + Enter.
'Hola, ciencia de datos!'
```

## 3. Comentarios en Python

Todo lo que sigue a `#` en una línea es un comentario: no se ejecuta, es solo una nota
para quien lea el código.

```python
# Esta línea es un comentario, no hace nada al ejecutar la celda
resultado = 3 + 3  # también puedes comentar al final de una línea
print(resultado)
```

## 4. Gestionar el archivo del notebook

- **Ejecutar todo**: `Runtime > Run all` (Colab) o `Cell > Run All` (Jupyter clásico).
- **Renombrar**: clic sobre el nombre del notebook en la parte superior.
- **Reiniciar el kernel**: si algo se comporta raro, `Runtime > Restart runtime`
  (Colab) es el equivalente a "apagar y volver a encender".
- **Guardar y descargar**: `File > Download > Download .ipynb`.

Una buena práctica es dejar, como primera celda de cualquier notebook nuevo, una
celda de "configuración":

```python
import sys
print("Versión de Python:", sys.version)

# Si el notebook depende de una librería que no viene preinstalada:
# !pip install nombre_de_la_libreria
```

## Resumen

- Shift + Enter ejecuta una celda; el estado (variables) se mantiene entre celdas.
- Los comentarios (`#`) no se ejecutan, son documentación para humanos.
- Reiniciar el kernel resuelve la mayoría de comportamientos extraños.
- Deja siempre una celda de configuración al principio de cada notebook.
