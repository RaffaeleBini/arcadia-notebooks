## 1. El patrón de una API

Cliente -> Petición (Request) -> Endpoint -> Respuesta (Response, normalmente en JSON).

Vamos a hacerlo real con una API pública gratuita que no requiere clave de acceso:
Open-Meteo, que devuelve el clima actual de cualquier coordenada.

```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
parametros = {
    "latitude": 40.71,
    "longitude": -74.01,
    "current_weather": "true",
}

respuesta = requests.get(url, params=parametros)
datos = respuesta.json()

print("Código de estado HTTP:", respuesta.status_code)
print("Temperatura actual en Nueva York:", datos["current_weather"]["temperature"], "°C")
```

## 2. Datasets y licencias

Un dataset puede ser tabular (CSV), jerárquico o de red/grafo. Existen catálogos
enormes de datos abiertos: data.gov, kaggle.com/datasets, data.worldbank.org, who.int.

Cada dataset tiene una licencia (Dominio Público, CC-BY, CDLA-Sharing, CDLA-Permissive...)
que define qué puedes hacer con él. El dataset meteorológico que usaremos en el
capítulo de cierre del curso está bajo licencia CDLA-Sharing.

```python
import pandas as pd

url_clima = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/fJKLoMKtgpDAl5MkA6rf2A/jfk-weather.csv"
clima = pd.read_csv(url_clima, parse_dates=["DATE"], low_memory=False)

print(f"Filas: {clima.shape[0]}, columnas: {clima.shape[1]}")
clima.head()
```

## Resumen

- Una API es un contrato de comunicación: pides algo a un endpoint concreto y
  recibes una respuesta (normalmente JSON) sin conocer los detalles internos.
- `requests.get()` es la forma estándar de consumir una API en Python.
- Cada dataset tiene una licencia; revísala antes de reutilizarlo en un proyecto real.
