## 1. NumPy: calcolo numerico veloce

NumPy lavora con *array*: collezioni di numeri su cui puoi applicare
operazioni matematiche a tutti gli elementi contemporaneamente, senza cicli.

```python
import numpy as np

horas_estudio = np.array([1, 2, 3, 4, 5])
nota_examen = np.array([50, 55, 65, 70, 85])

print("Promedio de horas:", np.mean(horas_estudio))
print("Nota máxima:", np.max(nota_examen))
print("Notas + 5 puntos de bonus:", nota_examen + 5)  # operazione vettorializzata
```

## 2. Pandas: la tabella (DataFrame)

Pandas costruisce, sopra NumPy, il suo pezzo forte: il `DataFrame`, una tabella con righe
e colonne etichettate.

```python
import pandas as pd

df = pd.DataFrame({"horas": horas_estudio, "nota": nota_examen})
print(df)

print("\nFilas donde la nota es mayor a 60:")
print(df[df["nota"] > 60])
```

## 3. Matplotlib: visualizzare

```python
import matplotlib.pyplot as plt

plt.scatter(df["horas"], df["nota"])
plt.xlabel("Horas de estudio")
plt.ylabel("Nota del examen")
plt.title("Relación entre horas de estudio y nota")
plt.show()
```

## 4. Scikit-learn: un primo modello

```python
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()
modelo.fit(df[["horas"]], df["nota"])

prediccion = modelo.predict([[6]])
print(f"Nota estimada con 6 horas de estudio: {prediccion[0]:.1f}")
```
