## 1. Por qué existe el control de versiones

Git registra todos los cambios de un proyecto para poder consultar el historial,
comparar versiones o volver atrás si algo sale mal. Es un sistema **distribuido**:
cada copia tiene el historial completo, no depende de un único servidor central.

Comandos esenciales: `git init/clone`, `git status`, `git add`, `git commit`,
`git log`, `git branch`, `git checkout`, `git merge`, `git reset`.

## 2. El ciclo básico: init, add, commit, log

```python
!mkdir -p /content/mi_proyecto
%cd /content/mi_proyecto
!git init
!git config user.email "estudiante@ejemplo.com"
!git config user.name "Estudiante del curso"
```

```python
# Creamos un archivo y revisamos qué detecta Git
!echo "print('hola mundo')" > script.py
!git status
```

```python
# Confirmamos el cambio en el historial (commit)
!git add script.py
!git commit -m "Primer commit: añade script.py"
!git log --oneline
```

## Resumen

- Git es un sistema de control de versiones distribuido.
- El ciclo básico es: `git init` (o `clone`) -> `git add` -> `git commit` -> `git log`.
- `git status` te dice en todo momento qué ha cambiado desde el último commit.
