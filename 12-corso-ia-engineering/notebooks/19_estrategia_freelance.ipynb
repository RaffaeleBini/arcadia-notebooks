## Celda 2 — Calcula tu day rate

Introduce tus números reales para calcular el precio mínimo al que puedes trabajar.

````python
# ── Introduce tus números ─────────────────────────────────────────────────────

FINANZAS = {
    # Coste de vida mensual (alquiler, comida, transporte, seguros, etc.)
    "coste_vida_mes_eur": 2500,          # TODO

    # Días que esperas facturar por mes (realista: 12-18 para empezar)
    "dias_facturables_mes": 15,          # TODO

    # Tasa impositiva estimada (autónomos España: ~30-35%)
    "tasa_impuestos": 0.32,

    # Coste de herramientas y suscripciones mensuales
    "herramientas_mes_eur": 150,         # GitHub Copilot, APIs, nube, etc.

    # Días de vacaciones al año
    "vacaciones_dias": 25,
}

# ── Cálculo ───────────────────────────────────────────────────────────────────

dias_laborables_mes = 22  # media
fraccion_vacaciones = 1 - FINANZAS["vacaciones_dias"] / 250  # 250 días laborables/año

# Ingresos brutos necesarios para cubrir coste de vida + impuestos + herramientas
coste_total_mes = (
    FINANZAS["coste_vida_mes_eur"]
    / (1 - FINANZAS["tasa_impuestos"])
    + FINANZAS["herramientas_mes_eur"]
)

# Días facturables ajustados a vacaciones
dias_fact_ajustados = FINANZAS["dias_facturables_mes"] * fraccion_vacaciones

day_rate_minimo = coste_total_mes / dias_fact_ajustados
day_rate_objetivo = day_rate_minimo * 1.5   # margen para crecer, formación, imprevistos
day_rate_senior = day_rate_minimo * 2.2

print("=" * 50)
print(" Tu calculadora de day rate")
print("=" * 50)
print(f"  Coste vida mensual:      €{FINANZAS['coste_vida_mes_eur']:,.0f}")
print(f"  Coste total (bruto):     €{coste_total_mes:,.0f}/mes")
print(f"  Días facturables/mes:    {dias_fact_ajustados:.1f} (ajustado vacaciones)")
print()
print(f"  Day rate MÍNIMO:         €{day_rate_minimo:,.0f}/día  ← no bajes de aquí")
print(f"  Day rate OBJETIVO:       €{day_rate_objetivo:,.0f}/día  ← apunta a esto")
print(f"  Day rate SENIOR:         €{day_rate_senior:,.0f}/día  ← meta a 2 años")
print()

# Equivalencia en proyecto tipo DocBot
dias_proyecto = 12  # discovery + desarrollo + despliegue
precio_proyecto_min = day_rate_minimo * dias_proyecto
precio_proyecto_obj = day_rate_objetivo * dias_proyecto
print(f"  Proyecto tipo DocBot ({dias_proyecto} días):")
print(f"    Mínimo:    €{precio_proyecto_min:,.0f}")
print(f"    Objetivo:  €{precio_proyecto_obj:,.0f}")
print()

# Ingresos anuales
ingresos_anuales_obj = day_rate_objetivo * FINANZAS["dias_facturables_mes"] * 11
print(f"  Ingresos anuales (objetivo, 11 meses):  €{ingresos_anuales_obj:,.0f}")
````
