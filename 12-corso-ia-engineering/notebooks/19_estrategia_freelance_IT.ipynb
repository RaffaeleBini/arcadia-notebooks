## Cella 2 — Calcola il tuo day rate

Inserisci i tuoi numeri reali per calcolare il prezzo minimo a cui puoi lavorare.

````python
# ── Inserisci i tuoi numeri ───────────────────────────────────────────────────

FINANZAS = {
    # Costo di vita mensile (affitto, cibo, trasporti, assicurazioni, ecc.)
    "coste_vida_mes_eur": 2500,          # TODO

    # Giorni che prevedi di fatturare al mese (realistico: 12-18 per iniziare)
    "dias_facturables_mes": 15,          # TODO

    # Aliquota fiscale stimata (liberi professionisti Spagna: ~30-35%)
    "tasa_impuestos": 0.32,

    # Costo di strumenti e abbonamenti mensili
    "herramientas_mes_eur": 150,         # GitHub Copilot, API, cloud, ecc.

    # Giorni di ferie all'anno
    "vacaciones_dias": 25,
}

# ── Calcolo ───────────────────────────────────────────────────────────────────

dias_laborables_mes = 22  # media
fraccion_vacaciones = 1 - FINANZAS["vacaciones_dias"] / 250  # 250 giorni lavorativi/anno

# Entrate lorde necessarie per coprire costo di vita + tasse + strumenti
coste_total_mes = (
    FINANZAS["coste_vida_mes_eur"]
    / (1 - FINANZAS["tasa_impuestos"])
    + FINANZAS["herramientas_mes_eur"]
)

# Giorni fatturabili corretti per le ferie
dias_fact_ajustados = FINANZAS["dias_facturables_mes"] * fraccion_vacaciones

day_rate_minimo = coste_total_mes / dias_fact_ajustados
day_rate_objetivo = day_rate_minimo * 1.5   # margine per crescere, formazione, imprevisti
day_rate_senior = day_rate_minimo * 2.2

print("=" * 50)
print(" La tua calcolatrice del day rate")
print("=" * 50)
print(f"  Costo di vita mensile:   €{FINANZAS['coste_vida_mes_eur']:,.0f}")
print(f"  Costo totale (lordo):    €{coste_total_mes:,.0f}/mese")
print(f"  Giorni fatturabili/mese: {dias_fact_ajustados:.1f} (corretto per le ferie)")
print()
print(f"  Day rate MINIMO:         €{day_rate_minimo:,.0f}/giorno  ← non scendere sotto")
print(f"  Day rate OBIETTIVO:      €{day_rate_objetivo:,.0f}/giorno  ← punta a questo")
print(f"  Day rate SENIOR:         €{day_rate_senior:,.0f}/giorno  ← meta a 2 anni")
print()

# Equivalenza in un progetto tipo DocBot
dias_proyecto = 12  # discovery + sviluppo + deploy
precio_proyecto_min = day_rate_minimo * dias_proyecto
precio_proyecto_obj = day_rate_objetivo * dias_proyecto
print(f"  Progetto tipo DocBot ({dias_proyecto} giorni):")
print(f"    Minimo:    €{precio_proyecto_min:,.0f}")
print(f"    Obiettivo: €{precio_proyecto_obj:,.0f}")
print()

# Entrate annuali
ingresos_anuales_obj = day_rate_objetivo * FINANZAS["dias_facturables_mes"] * 11
print(f"  Entrate annuali (obiettivo, 11 mesi):  €{ingresos_anuales_obj:,.0f}")
````
