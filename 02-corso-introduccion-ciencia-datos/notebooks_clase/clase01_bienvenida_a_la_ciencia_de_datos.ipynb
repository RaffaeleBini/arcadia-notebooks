## 1. ¿Qué es la ciencia de datos?

La ciencia de datos es **el estudio de los datos para entender el mundo** — o, dicho de forma más evocadora, "el arte de descubrir insights y tendencias escondidas detrás de los datos".

Importante: analizar datos no es una actividad nueva (llevamos siglos haciéndolo). Lo que **sí** es nuevo es la cantidad de datos disponibles y la potencia de cómputo para procesarlos.

### El proceso de un/a data scientist
1. Clarificar el problema
2. Recolectar datos
3. Analizar
4. Reconocer patrones
5. Contar la historia (storytelling)
6. Visualizar

## 2. Las tres cualidades vitales (Murtaza Haider)

- **Curiosidad**: explorar datos y hacer preguntas.
- **Argumentación**: explicar hallazgos de forma convincente.
- **Buen juicio**: saber por dónde empezar.

Un/a data scientist **no es solo un estadístico ni solo un programador**: es una mezcla de ambos, más comunicación. Vienen de disciplinas muy diversas: economía, ingeniería, medicina...

### Demo en vivo: descubriendo un patrón escondido

Vamos a simular datos de una tienda ficticia (ventas de paraguas y helados según si llovió o no) y vamos a hacer exactamente el paso 4 del proceso: **reconocer patrones**.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_dias = 200

llovio = np.random.choice([0, 1], size=n_dias, p=[0.7, 0.3])
ventas_paraguas = np.random.randint(2, 6, size=n_dias) + llovio * np.random.randint(8, 16, size=n_dias)
ventas_helados = np.random.randint(15, 25, size=n_dias) - llovio * np.random.randint(5, 12, size=n_dias)
ventas_helados = np.clip(ventas_helados, 0, None)

tienda = pd.DataFrame({
    "llovio": llovio,
    "ventas_paraguas": ventas_paraguas,
    "ventas_helados": ventas_helados,
})

resumen = tienda.groupby("llovio")[["ventas_paraguas", "ventas_helados"]].mean()
print(resumen)
```

Fíjate cómo, con `groupby`, comparamos los días de lluvia contra los días sin lluvia de un solo vistazo. Ese es el tipo de hallazgo — sencillo pero accionable — que un/a data scientist convierte en una historia para el negocio.

## 3. Un día en la vida real: casos concretos

- **Transporte en Toronto**: el Dr. Murtaza Haider explicó por qué 500.000 personas se quejaron del transporte público encontrando una relación con eventos climáticos inesperados.
- **Floraciones de algas**: redes neuronales para predecir toxicidad del agua y proteger el ecosistema.
- **Motores de recomendación**: Norman White (Stern School of Business) simplificó problemas complejos entre departamentos.

Herramienta estadística clave que verás mucho: la **regresión** — muestra la relación probable entre dos variables (ej. distancia recorrida en coche y gasolina consumida).

### Demo en vivo: cruzando dos fuentes de datos (clima y quejas)

```python
np.random.seed(7)
n_dias = 90

evento_climatico = np.random.choice([0, 1], size=n_dias, p=[0.8, 0.2])
quejas = np.random.randint(50, 150, size=n_dias) + evento_climatico * np.random.randint(200, 400, size=n_dias)

transporte = pd.DataFrame({
    "evento_climatico": evento_climatico,
    "quejas": quejas,
})

print(transporte.groupby("evento_climatico")["quejas"].describe())

correlacion = transporte["evento_climatico"].corr(transporte["quejas"])
print(f"\nCorrelación evento climático vs. quejas: {correlacion:.2f}")
```

Un valor de correlación cercano a 1 confirma numéricamente lo que el Dr. Haider encontró en la vida real: existe una relación fuerte entre el evento climático y el número de quejas ese mismo día.

## 4. ¿Estadística o ciencia de datos?

Hal Varian (economista jefe de Google) dijo que el trabajo "sexy" de la próxima década sería el de estadístico — afirmación polémica. La diferencia clave: un/a data scientist usa estadística, **pero también** programación, bases de datos y comunicación (storytelling). Rachel Schutt (chief data scientist de News Corp) lo resume así: la curiosidad es lo que hace excepcional a un/a data scientist, que es una mezcla de científico de la computación, ingeniero de software y estadístico.

**Idea central de la clase**: la ciencia de datos no es solo una profesión — es un viaje de exploración, innovación y narración de historias.
