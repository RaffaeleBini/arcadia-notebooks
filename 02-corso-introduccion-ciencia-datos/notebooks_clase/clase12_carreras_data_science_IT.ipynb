## Il mito dell'unicorno

Iniziamo smontando un mito. Nel settore si parla molto di un profilo quasi impossibile: qualcuno con esperienza profonda di business, capacità di analisi tecnica completa (dati strutturati e non strutturati) e una narrazione eccellente, tutto in una sola persona. Lo si chiama, con ironia, l'**"unicorno"**: tutti ne parlano, quasi nessuno l'ha mai visto davvero.

Il consiglio reale di chi assume è un altro: assumere per **affinità culturale e passione**, non inseguire la persona perfetta, e costruire **team** dove quegli elementi (business, tecnica, comunicazione) sono distribuiti tra più persone.

Se stai iniziando, questa è una buona notizia: non devi essere tutto. Devi essere un pezzo prezioso di una squadra.

## L'ordine di priorità nelle assunzioni

Un'intervista con professionisti del settore ha rivelato un ordine che sorprende:

1. **Curiosità**
2. **Senso dell'umorismo / affinità personale**
3. **Competenze tecniche** — all'ultimo posto

Perché le competenze tecniche vengono per ultime? Perché sono le più facili da insegnare. Un linguaggio di programmazione si impara in mesi; la curiosità genuina, molto più lentamente. Questo non significa che la tecnica non conti — serve un minimo reale — ma che, tra due candidati con un livello tecnico ragionevole, vince chi ha più curiosità e migliore affinità.

## Il percorso tecnico in tre passi

Per i ruoli big data, il percorso tecnico si riassume in:

1. **Archiviare i dati** (database relazionali, NoSQL, warehouse — lezioni 9-11).
2. **Manipolare i dati** (pulire, trasformare, preparare).
3. **Applicare algoritmi** (machine learning — lezioni 5 e 6).

Strumenti tipici: R/Stata/Python per l'analitica predittiva tradizionale, Python per dati non strutturati, Hadoop/Spark per il big data (lezione 4).

## Comunicazione: la vista dalla valle alla montagna

Il Dr. Murtaza Haider usa questa metafora: non mostrare solo il sentiero pianeggiante della tua analisi — porta il tuo pubblico fino al belvedere da cui all'improvviso si vede l'intera valle. Quel momento di "aha" è ciò che la gente ricorda. Su questo approfondiamo nella lezione 13.

## Il mercato del lavoro, in cifre

- Global Industry Analysts Inc. ha previsto che il mercato delle piattaforme di scienza dei dati avrebbe raggiunto 314.800 milioni di dollari nel 2025, con un CAGR del 38,2%.
- Il McKinsey Global Institute avvertì di una carenza di talento analitico per il 2018.

## Se inizi da zero

1. Impara a programmare.
2. Impara matematica, probabilità e statistica — quanto basta, non un'intera laurea.
3. Gioca e costruisci cose con quanto sopra.

Per studenti delle superiori: SQL, database, e qualsiasi corso di informatica disponibile. Pensa a una fiera della scienza: vince chi ha un'ipotesi chiara messa alla prova con cura, non chi ha la calligrafia più bella.

## Demo dal vivo: autovalutazione delle competenze

Costruiamo un grafico a barre orizzontale che riassuma, a colpo d'occhio, un esempio di autovalutazione di competenze di scienza dei dati. Questo è esattamente ciò che adatterai tu stesso/a nella pratica di questa lezione.

```python
import matplotlib.pyplot as plt

# Elenco di competenze rilevanti secondo questo capitolo, con un'autovalutazione
# di esempio su una scala da 1 (nulla) a 5 (padronanza alta).
habilidades = ["Python", "SQL", "Estadística", "Comunicación", "Curiosidad"]
nivel = [3, 2, 3, 4, 5]

fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(habilidades, nivel, color="steelblue")
ax.set_xlabel("Livello autopercepito (1 = nulla, 5 = padronanza alta)")
ax.set_title("Esempio di autovalutazione di competenze di scienza dei dati")
ax.set_xlim(0, 5)
ax.axvline(3, color="gray", linestyle="--", linewidth=1)

plt.tight_layout()
plt.show()
```

Nota che la linea tratteggiata al livello 3 ci dà un riferimento rapido: qualsiasi barra che resta sotto quella linea è una buona candidata su cui investire tempo di studio nelle prossime settimane. Questo stesso grafico, con i tuoi numeri, è il tuo vero punto di partenza per un piano di miglioramento.
