## Tre cerchi, uno dentro l'altro

Quando qualcuno dice "usiamo l'IA" e poi "abbiamo addestrato un modello di deep learning" come se fossero sinonimi, sta mescolando cose di portata molto diversa. Pensali come cerchi concentrici:

**Intelligenza Artificiale (IA)** ⊃ **Machine Learning (ML)** ⊃ **Deep Learning (DL)**

Tutto il Deep Learning è Machine Learning. Tutto il Machine Learning è Intelligenza Artificiale. Ma non viceversa: c'è moltissima IA che non è Machine Learning (ad esempio, un sistema esperto basato su regole scritte a mano), e molto Machine Learning che non arriva a essere Deep Learning.

```python
# Un modo semplice ma efficace di visualizzare la relazione di
# contenimento con insiemi di Python: ogni insieme "più specifico"
# è un sottoinsieme del precedente.
DL = {"redes neuronales convolucionales", "redes neuronales recurrentes", "transformers"}
ML = DL | {"regresion", "arboles de decision", "knn"}
IA = ML | {"reglas expertas", "busqueda", "logica difusa"}

print("DL è sottoinsieme di ML?", DL.issubset(ML))
print("ML è sottoinsieme di IA?", ML.issubset(IA))
print("Elementi che sono in IA ma NON in ML:", IA - ML)
```

## Machine Learning: imparare dagli esempi, non dalle regole

Nella programmazione tradizionale, tu scrivi le regole e il computer le esegue. Nel Machine Learning, dai all'algoritmo molti esempi già risolti (dati di addestramento), ed è l'algoritmo stesso a scoprire il pattern, generalizzandolo per casi nuovi.

## Deep Learning: reti neurali a strati

Un sottoinsieme specializzato del ML che usa **reti neurali artificiali** a strati per simulare, in modo semplificato, come il cervello prende decisioni. Differenza chiave: molti algoritmi di ML "classico" si stabilizzano (fanno *plateau*) dandogli più dati; il Deep Learning tende a **migliorare** quanti più dati riceve.

Una rete neurale è una raccolta di "neuroni" organizzati in strati che trasformano l'input passo dopo passo e si addestrano iterativamente. Quando ci sono molti strati, si dice che la rete è "profonda" — da qui il nome. Questa tecnica è rimasta quasi 15 anni in disuso per il suo costo computazionale, finché le **GPU** (con centinaia di core in parallelo — una GPU equivale approssimativamente a 600 core) l'hanno resa di nuovo fattibile.

## L'IA e la Scienza dei Dati sono la stessa cosa?

No. Non c'è relazione di sottoinsieme tra loro. La **Scienza dei Dati** è il processo di estrarre conoscenza da grandi volumi di dati (mescola matematica, statistica, visualizzazione, programmazione). L'**IA** è tutto ciò che permette alle macchine di imparare a risolvere problemi da sole. Si intrecciano costantemente: la scienza dei dati usa tecniche di IA (incluso ML e DL) come strumenti, ma non è un sottoinsieme dell'IA né viceversa.

## Demo dal vivo: regole fisse vs. un modello che impara

Risolviamo lo STESSO problema in due modi diversi: classificare fiori del dataset `iris` (incluso in scikit-learn, senza download) usando solo due misure del petalo.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

iris = load_iris()

# Teniamo 2 specie (0 e 1) e 2 colonne: lunghezza e larghezza del petalo
mascara = iris.target != 2
X = iris.data[mascara][:, [2, 3]]
y = iris.target[mascara]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Esempi di addestramento:", len(X_train))
print("Esempi di test:", len(X_test))
```

```python
# --- Approccio 1: regole fisse, scritte a mano da noi ---
def clasificar_con_reglas(largo_petalo, ancho_petalo):
    if largo_petalo < 2.5:
        return 0  # setosa
    return 1  # versicolor

predicciones_reglas = [clasificar_con_reglas(l, a) for l, a in X_test]
aciertos_reglas = np.mean(predicciones_reglas == y_test)
print(f"Regole fisse scritte a mano: {aciertos_reglas:.0%} di accuratezza")

# --- Approccio 2: un modello che IMPARA dai dati ---
modelo = KNeighborsClassifier(n_neighbors=3)
modelo.fit(X_train, y_train)  # qui "impara": memorizza gli esempi e la loro geometria
aciertos_modelo = modelo.score(X_test, y_test)
print(f"Modello addestrato (KNN): {aciertos_modelo:.0%} di accuratezza")
```

Su questo dataset così "facile" entrambi gli approcci possono indovinare molto, perché le specie sono ben separate. Ma nota la differenza di processo: noi abbiamo **inventato** la soglia `2.5` guardando i dati a occhio; il modello KNN non ha mai visto nessuna regola — solo esempi già etichettati, e ha scoperto da solo come separarli confrontando ogni nuovo fiore con quelli più simili che già conosceva. Nei problemi reali, con decine di variabili, scrivere regole a mano smette di essere praticabile molto prima che addestrare un modello.

## Riepilogo della lezione

- IA ⊃ ML ⊃ DL: sono insiemi annidati, non sinonimi.
- Il ML impara pattern da dati di esempio; non segue regole scritte a mano.
- Il DL usa reti neurali a strati e migliora (invece di stabilizzarsi) quanti più dati riceve; le GPU l'hanno reso di nuovo fattibile.
- La Scienza dei Dati e l'IA sono campi distinti che si sostengono a vicenda, non uno sottoinsieme dell'altro.
- Abbiamo confrontato dal vivo un classificatore a regole fisse contro un modello KNN addestrato con dati, sullo stesso problema.
