## 1. Cos'è la scienza dei dati?

La scienza dei dati è **lo studio dei dati per capire il mondo** — o, detto in modo più evocativo, "l'arte di scoprire insight e tendenze nascoste dietro i dati".

Importante: analizzare dati non è un'attività nuova (lo facciamo da secoli). Ciò che **è** nuovo è la quantità di dati disponibili e la potenza di calcolo per elaborarli.

### Il processo di un/una data scientist
1. Chiarire il problema
2. Raccogliere dati
3. Analizzare
4. Riconoscere pattern
5. Raccontare la storia (storytelling)
6. Visualizzare

## 2. Le tre qualità vitali (Murtaza Haider)

- **Curiosità**: esplorare i dati e fare domande.
- **Argomentazione**: spiegare le scoperte in modo convincente.
- **Buon giudizio**: sapere da dove iniziare.

Un/una data scientist **non è solo uno statistico né solo un programmatore**: è un misto di entrambi, più comunicazione. Provengono da discipline molto diverse: economia, ingegneria, medicina...

### Demo dal vivo: scoprire un pattern nascosto

Simuliamo dati di un negozio fittizio (vendite di ombrelli e gelati a seconda che abbia piovuto o no) e facciamo esattamente il passo 4 del processo: **riconoscere i pattern**.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_dias = 200

llovio = np.random.choice([0, 1], size=n_dias, p=[0.7, 0.3])
ventas_paraguas = np.random.randint(2, 6, size=n_dias) + llovio * np.random.randint(8, 16, size=n_dias)
ventas_helados = np.random.randint(15, 25, size=n_dias) - llovio * np.random.randint(5, 12, size=n_dias)
ventas_helados = np.clip(ventas_helados, 0, None)

tienda = pd.DataFrame({
    "llovio": llovio,
    "ventas_paraguas": ventas_paraguas,
    "ventas_helados": ventas_helados,
})

resumen = tienda.groupby("llovio")[["ventas_paraguas", "ventas_helados"]].mean()
print(resumen)
```

Nota come, con `groupby`, confrontiamo i giorni di pioggia contro i giorni senza pioggia in un solo colpo d'occhio. Questo è il tipo di scoperta — semplice ma praticabile — che un/una data scientist trasforma in una storia per il business.

## 3. Un giorno nella vita reale: casi concreti

- **Trasporto a Toronto**: il Dr. Murtaza Haider ha spiegato perché 500.000 persone si sono lamentate del trasporto pubblico trovando una relazione con eventi climatici inaspettati.
- **Fioriture di alghe**: reti neurali per prevedere la tossicità dell'acqua e proteggere l'ecosistema.
- **Motori di raccomandazione**: Norman White (Stern School of Business) ha semplificato problemi complessi tra dipartimenti.

Strumento statistico chiave che vedrai molto: la **regressione** — mostra la relazione probabile tra due variabili (es. distanza percorsa in auto e benzina consumata).

### Demo dal vivo: incrociare due fonti di dati (clima e reclami)

```python
np.random.seed(7)
n_dias = 90

evento_climatico = np.random.choice([0, 1], size=n_dias, p=[0.8, 0.2])
quejas = np.random.randint(50, 150, size=n_dias) + evento_climatico * np.random.randint(200, 400, size=n_dias)

transporte = pd.DataFrame({
    "evento_climatico": evento_climatico,
    "quejas": quejas,
})

print(transporte.groupby("evento_climatico")["quejas"].describe())

correlacion = transporte["evento_climatico"].corr(transporte["quejas"])
print(f"\nCorrelazione evento climatico vs. reclami: {correlacion:.2f}")
```

Un valore di correlazione vicino a 1 conferma numericamente ciò che il Dr. Haider ha trovato nella vita reale: esiste una relazione forte tra l'evento climatico e il numero di reclami quello stesso giorno.

## 4. Statistica o scienza dei dati?

Hal Varian (capo economista di Google) ha detto che il lavoro "sexy" del decennio successivo sarebbe stato quello dello statistico — un'affermazione controversa. La differenza chiave: un/una data scientist usa la statistica, **ma anche** programmazione, database e comunicazione (storytelling). Rachel Schutt (chief data scientist di News Corp) lo riassume così: la curiosità è ciò che rende eccezionale un/una data scientist, che è una miscela di informatico, ingegnere del software e statistico.

**Idea centrale della lezione**: la scienza dei dati non è solo una professione — è un viaggio di esplorazione, innovazione e narrazione.

- Curiosità, argomentazione e buon giudizio sono le tre qualità vitali secondo Murtaza Haider.
- Un/una data scientist non è solo statistico né solo programmatore: è entrambi + comunicazione.
