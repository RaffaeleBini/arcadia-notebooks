# Clase 13: Comunicar resultados con storytelling

**Objetivos de aprendizaje:**
- Explicar por qué una narrativa persuade más que una tabla de números.
- Estructurar un informe de resultados.
- Construir una visualización clara con Matplotlib.

**Duración estimada:** 90 min

## ¿Por qué importa esto?

Puedes pasar semanas haciendo un análisis brillante y aun así fracasar por completo si no sabes comunicarlo. Los datos, por sí solos, no convencen a nadie — lo que convence es la historia que construyes alrededor de ellos. Recuerda el proceso del capítulo 1: no termina en "analizar", termina en "contar la historia" y "visualizar".

## ¿Cuánto debe medir un informe?

No hay una respuesta única:
- Académico: 1.000 a 7.000 palabras.
- Negocio/consultoría: desde menos de 1.500 palabras hasta cientos de páginas.

Firmas como McKinsey y Deloitte han construido buena parte de su reputación sobre informes basados en análisis de datos.

## Un ejemplo real: el informe de Deloitte

"United States Economic Forecast" (Deloitte University Press, diciembre 2014, 24 páginas):

- Abre con una afirmación que capta la atención (la economía está más sana de lo que se percibe).
- Cita a Voltaire para enmarcar el tono.
- Usa gráficos de series temporales del PIB (contracción y recuperación de la Gran Recesión).
- Presenta **cuatro escenarios futuros**, no una única predicción.
- Conecta con Thomas Piketty y el debate sobre desigualdad de ingresos.
- Relega el detalle técnico (4 tablas de datos) a un apéndice.

**Argumento central:** esa narrativa persuade mucho más que esos mismos 8 gráficos y 4 tablas presentados sueltos, como diapositivas sin hilo conductor.

## Planifica el final antes de empezar

Recomendación clave: **planifica el informe final antes de empezar el análisis**, no después. Así, cada decisión durante el análisis (qué datos priorizar, qué gráficos preparar) está guiada por el objetivo de comunicación.

Un buen informe comunica: beneficio para el lector, objetivos claros, importancia del tema, contexto, aplicabilidad práctica y direcciones futuras.

La estructura depende del propósito y la longitud: informes breves van directos al grano; informes extensos construyen el argumento progresivamente y siguen un formato más prescrito (portada, índice, resumen ejecutivo, contenido, agradecimientos). Dato curioso: hasta candidatos a doctorado olvidan a veces la portada.

## Demo en vivo: de "mostrar datos" a "contar una historia"

Vamos a partir de una tabla pequeña hecha a mano —una serie temporal con varios escenarios, al estilo del informe de Deloitte— y comparar dos formas de presentarla: en crudo, y como una historia visual con una anotación que señala el punto clave.

```python
import matplotlib.pyplot as plt

# Tabla pequeña hecha a mano: "crecimiento económico proyectado" bajo tres escenarios,
# inspirada en el tipo de dato de un informe de previsión económica.
anios = [2021, 2022, 2023, 2024, 2025]
escenario_pesimista  = [1.0, 0.5, 0.3, 0.8, 1.2]
escenario_base       = [1.0, 1.8, 2.1, 2.4, 2.6]
escenario_optimista  = [1.0, 2.5, 3.4, 4.0, 4.5]

# Paso 1: "mostrar datos" en crudo
print("Escenario base:", escenario_base)
```

```python
# Paso 2: "contar una historia" con el mismo dato
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(anios, escenario_pesimista, label="Escenario pesimista", color="firebrick", linestyle="--")
ax.plot(anios, escenario_base, label="Escenario base", color="steelblue", linewidth=2.5)
ax.plot(anios, escenario_optimista, label="Escenario optimista", color="seagreen", linestyle="--")

ax.set_title("Crecimiento económico proyectado bajo tres escenarios")
ax.set_xlabel("Año")
ax.set_ylabel("Crecimiento proyectado (%)")
ax.legend(loc="upper left")

# La anotación señala explícitamente el punto que queremos que la audiencia recuerde
ax.annotate(
    "La recuperación se acelera\na partir de 2023",
    xy=(2023, escenario_base[2]),
    xytext=(2021.3, 3.6),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=10,
)

plt.tight_layout()
plt.show()
```

Compara las dos celdas anteriores: la primera es información correcta pero sin foco — el equivalente a los "8 gráficos y 4 tablas sin hilo narrativo". La segunda organiza los tres escenarios de un vistazo, y la llamada a `plt.annotate()` es la que convierte el gráfico en una historia: en lugar de esperar a que la audiencia detecte el cambio importante por sí misma, se lo señalamos de forma explícita — igual que Deloitte abrió su informe con una afirmación en vez de dejar que el lector llegara solo a esa conclusión 20 páginas después.

## Resumen de la clase

- Los datos no persuaden por sí solos — persuade la narrativa alrededor de ellos.
- Un buen informe combina: afirmación inicial fuerte, contexto/referencias, visualizaciones que resumen la historia, reconocimiento de la incertidumbre, y detalle técnico relegado a un apéndice.
- Planifica el informe final antes de analizar, no después.
- Título, etiquetas de ejes y sobre todo `plt.annotate()` son lo que convierte "mostrar un gráfico" en "contar una historia con un gráfico".
