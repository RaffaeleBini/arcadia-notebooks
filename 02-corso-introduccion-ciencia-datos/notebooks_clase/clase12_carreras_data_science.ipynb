## El mito del unicornio

Empecemos por desmontar un mito. En el sector se habla mucho de un perfil casi imposible: alguien con experiencia profunda de negocio, capacidad de análisis técnico completo (datos estructurados y no estructurados) y una narrativa excelente, todo en una sola persona. A eso se le llama, con sorna, el **"unicornio"**: todo el mundo habla de él, casi nadie lo ha visto de verdad.

La recomendación real de quienes contratan es otra: contratar por **encaje cultural y pasión**, no perseguir a la persona perfecta, y construir **equipos** donde esas piezas (negocio, técnica, comunicación) estén repartidas entre varias personas.

Si estás empezando, esto es una buena noticia: no necesitas serlo todo. Necesitas ser una pieza valiosa de un equipo.

## El orden de prioridad al contratar

Una entrevista con profesionales del sector reveló un orden que sorprende:

1. **Curiosidad**
2. **Sentido del humor / encaje personal**
3. **Habilidades técnicas** — en último lugar

¿Por qué las técnicas van últimas? Porque son las más fáciles de enseñar. Un lenguaje de programación se aprende en meses; la curiosidad genuina, mucho más despacio. Esto no significa que la técnica no importe — hace falta un mínimo real — sino que, entre dos candidatos con nivel técnico razonable, gana quien tiene más curiosidad y mejor encaje.

## El camino técnico en tres pasos

Para roles de big data, el camino técnico se resume en:

1. **Almacenar datos** (bases relacionales, NoSQL, warehouses — clases 9 a 11).
2. **Manipular datos** (limpiar, transformar, preparar).
3. **Aplicar algoritmos** (machine learning — clases 5 y 6).

Herramientas típicas: R/Stata/Python para analítica predictiva tradicional, Python para datos no estructurados, Hadoop/Spark para big data (clase 4).

## Si empiezas de cero

1. Aprende a programar.
2. Aprende matemáticas, probabilidad y estadística — lo suficiente, no una carrera entera.
3. Juega y construye cosas con lo anterior.

Para estudiantes de instituto: SQL, bases de datos, y cualquier curso de ciencias de la computación disponible. Piensa en una feria de ciencias: gana quien tiene una hipótesis clara puesta a prueba con cuidado, no quien tiene la caligrafía más bonita.

## Demo en vivo: autoevaluación de habilidades

Vamos a construir un gráfico de barras horizontal que resuma, de un vistazo, un ejemplo de autoevaluación de habilidades de ciencia de datos. Esto es exactamente lo que vas a adaptar tú mismo/a en la práctica de esta clase.

```python
import matplotlib.pyplot as plt

# Lista de habilidades relevantes según este capítulo, con una autoevaluación
# de ejemplo en una escala de 1 (nada) a 5 (dominio alto).
habilidades = ["Python", "SQL", "Estadística", "Comunicación", "Curiosidad"]
nivel = [3, 2, 3, 4, 5]

fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(habilidades, nivel, color="steelblue")
ax.set_xlabel("Nivel autopercibido (1 = nada, 5 = dominio alto)")
ax.set_title("Ejemplo de autoevaluación de habilidades de ciencia de datos")
ax.set_xlim(0, 5)
ax.axvline(3, color="gray", linestyle="--", linewidth=1)

plt.tight_layout()
plt.show()
```

Fíjate en que la línea discontinua en el nivel 3 nos da una referencia rápida: cualquier barra que se queda corta respecto a esa línea es una buena candidata para invertir tiempo de estudio en las próximas semanas. Este mismo gráfico, con tus propios números, es tu punto de partida real para un plan de mejora.
