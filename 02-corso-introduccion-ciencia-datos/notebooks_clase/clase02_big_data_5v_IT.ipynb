## 1. Perché parliamo di "Big Data"?

Ogni volta che usi il cellulare, uno smartwatch o un'app di streaming lasci una traccia digitale. Da quell'accumulo massiccio e costante di dati nasce il termine **Big Data**.

Definizione di Ernst & Young: *"Il Big Data si riferisce ai volumi dinamici, grandi e disparati di dati creati da persone, strumenti e macchine. Richiede tecnologia nuova, innovativa e scalabile per raccogliere, archiviare ed elaborare analiticamente quella quantità di dati al fine di ottenere insight di business in tempo reale relativi a consumatori, rischio, guadagni, prestazioni, produttività e valore per gli azionisti"*.

Non esiste un'unica definizione universale — ma quasi tutte concordano su 5 elementi comuni: **le 5V**.

## 2. Le 5V

| V | Cosa misura | Esempio |
|---|---|---|
| **Volume** | La scala dei dati | ~7 miliardi di persone generano ~2,5 trilioni di byte/giorno |
| **Velocità** | La rapidità di accumulo | Ogni 60 secondi vengono caricate ore di video su YouTube |
| **Varietà** | La diversità di formati/fonti | Testo, immagini, video, wearable, IoT |
| **Veridicità** | La qualità e affidabilità | ~80% dei dati è non strutturato |
| **Valore** | La capacità di trasformare i dati in qualcosa di utile | Economico, medico, sociale... |

Definizione alternativa e pratica: il Big Data è ciò che non puoi più gestire con sistemi di database tradizionali, per il suo volume e velocità combinati. Origine storica: Google (Larry Page e Sergey Brin) dovette inventare tecnologia nuova per l'algoritmo PageRank e per archiviare l'intero web — da qui nasce ciò che poi avrebbe ispirato Hadoop (prossima lezione).

### Demo dal vivo: misurare Volume, Varietà e Veridicità

Simuliamo 1000 eventi di sensori di una città intelligente e misuriamo tre delle 5V direttamente sul DataFrame.

```python
import pandas as pd
import numpy as np

np.random.seed(3)
n_eventos = 1000

tipos_sensor = np.random.choice(["trafico", "clima", "wearable", "camara"], size=n_eventos)
medida = np.random.normal(loc=50, scale=15, size=n_eventos)

# Simuliamo un po' di "rumore" reale: valori mancanti ed errati
idx_faltantes = np.random.choice(n_eventos, size=int(n_eventos * 0.05), replace=False)
medida[idx_faltantes] = np.nan
idx_erroneos = np.random.choice(n_eventos, size=int(n_eventos * 0.02), replace=False)
medida[idx_erroneos] = -999

eventos = pd.DataFrame({"tipo_sensor": tipos_sensor, "medida": medida})

print("VOLUME (num. di eventi):", len(eventos))
print("\nVARIETÀ (tipi di sensore):")
print(eventos["tipo_sensor"].value_counts())

faltantes = eventos["medida"].isna().sum()
erroneos = (eventos["medida"] == -999).sum()
utilizables = len(eventos) - faltantes - erroneos
print(f"\nVERIDICITÀ: {utilizables}/{len(eventos)} record utilizzabili ({utilizables/len(eventos):.1%})")
```

Nota come il Volume (`len`), la Varietà (`value_counts`) e la Veridicità (proporzione di dati puliti) si possano misurare con strumenti di pandas che già conosci — ma su scala reale (milioni o miliardi di righe) queste stesse operazioni richiedono calcolo distribuito, argomento della prossima lezione.

## 3. Perché gli strumenti tradizionali non bastano?

Gli strumenti convenzionali di analisi (pensa ad aprire un file gigante in un foglio di calcolo) **non sono percorribili** alla scala del Big Data. Servono strumenti di **calcolo distribuito** — distribuire il lavoro tra molti computer contemporaneamente. È di questo che tratta l'ecosistema Hadoop, Hive e Spark, che vedremo più avanti nel corso.

- Una definizione pratica: il Big Data è ciò che non entra più in un sistema di database tradizionale.
- L'80% dei dati generati oggi è non strutturato, il che complica il garantire la Veridicità.
- Gli strumenti convenzionali non scalano a questo volume; servono strumenti di calcolo distribuito.
