## 1. Da analizzare dati a creare dati

Tutto ciò che abbiamo visto finora —regressione, classificazione, machine learning in generale— parte da dati che **già esistono** e cerca pattern in essi. L'IA generativa ribalta il problema: può una macchina **creare** dati nuovi che non esistevano, ma che siano plausibili?

Esempi che sicuramente già conosci senza chiamarli con il loro nome tecnico:
- Un chatbot che ti scrive un paragrafo coerente.
- Un'immagine generata a partire da una descrizione testuale.
- Musica o codice generati automaticamente.

Tutto questo è **IA generativa**: un sottoinsieme dell'IA incentrato sul produrre dati nuovi che imitano le creazioni umane.

## 2. Come funziona (a grandi linee)

Si appoggia su modelli di deep learning addestrati per generare, non solo per predire:

- **GAN (Reti Generative Avversarie)**: due reti che competono — una genera dati falsi, l'altra cerca di individuarli. Competendo, entrambe migliorano.
- **VAE (Autoencoder Variazionali)**: imparano a comprimere dati reali e a ricostruirli, generando variazioni nuove e plausibili.

Entrambi gli approcci imparano la **distribuzione** dei dati reali e generano istanze nuove che la rispettano, senza copiare esempi esistenti.

## 3. L'uso che ci interessa di più: dati sintetici

Addestrare e testare un modello richiede molti dati, e a volte non ci sono abbastanza osservazioni reali (sono costose, scarse, o protette dalla privacy).

I data scientist usano l'IA generativa per creare **dati sintetici**: dati artificiali con proprietà simili a quelle reali (stessa distribuzione, stessi raggruppamenti), che si combinano con i dati reali per addestrare e testare modelli con maggiore robustezza.

Non serve una GAN per capire l'idea di fondo. Se conosci la media e la deviazione standard di una variabile reale, puoi generare osservazioni nuove che seguano quella stessa distribuzione. Facciamolo dal vivo.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes

# Seme fisso perché il risultato sia riproducibile in classe
np.random.seed(42)

# Dataset reale, incluso in scikit-learn (non si scarica nulla da internet)
diabetes = load_diabetes()
print("Forma del dataset:", diabetes.data.shape)
print("Nomi delle colonne:", diabetes.feature_names)
```

```python
# Teniamo una colonna reale: l'indice di massa corporea (bmi)
bmi_real = diabetes.data[:, 2]

# Passo 1: misuriamo la distribuzione reale
media_real = bmi_real.mean()
desviacion_real = bmi_real.std()

print(f"Media reale:      {media_real:.4f}")
print(f"Deviazione reale: {desviacion_real:.4f}")
print(f"N. di osservazioni reali: {len(bmi_real)}")
```

```python
# Passo 2: generiamo dati sintetici con la stessa distribuzione
n_sinteticos = 1000
bmi_sintetico = np.random.normal(media_real, desviacion_real, n_sinteticos)

print(f"Media sintetica:      {bmi_sintetico.mean():.4f}")
print(f"Deviazione sintetica: {bmi_sintetico.std():.4f}")
print(f"N. di osservazioni sintetiche: {len(bmi_sintetico)}")
```

```python
# Passo 3: confrontiamo visivamente entrambe le distribuzioni
plt.figure(figsize=(8, 5))
plt.hist(bmi_real, bins=25, alpha=0.6, density=True, label="Dati reali")
plt.hist(bmi_sintetico, bins=25, alpha=0.6, density=True, label="Dati sintetici")
plt.title("BMI reale vs. BMI sintetico")
plt.xlabel("Indice di massa corporea (normalizzato)")
plt.ylabel("Densità")
plt.legend()
plt.show()
```

Nota come le due campane si sovrappongono quasi completamente: questa è evidenza visiva che i dati sintetici "si comportano" come quelli reali, anche se nessuna di quelle 1.000 osservazioni corrisponde a un paziente reale.

Questa è, in miniatura, la stessa logica usata da un data scientist quando ha bisogno di ampliare un dataset troppo piccolo per addestrare un modello: imparare la distribuzione del reale e generare più esempi che la rispettino.

## 4. Altre applicazioni dell'IA generativa

- Generazione di testo di tipo umano (GPT-3) per chatbot e creazione di contenuti.
- Sintesi di immagini mediche per la formazione dei professionisti sanitari.
- Arte generata, ambienti e livelli di videogiochi, design di moda.
- **Automazione del codice**: generare e testare automaticamente codice di modelli analitici.
- **Generazione di insight e report** che si aggiornano man mano che evolvono i dati (esempio: IBM Cognos Analytics, con un assistente in linguaggio naturale).

- L'IA generativa **crea** dati nuovi invece di analizzare solo quelli esistenti.
- Si appoggia su modelli come GAN e VAE, che imparano la distribuzione dei dati reali.
- Uno dei suoi usi più pratici per un data scientist: generare **dati sintetici** quando mancano dati reali sufficienti.
