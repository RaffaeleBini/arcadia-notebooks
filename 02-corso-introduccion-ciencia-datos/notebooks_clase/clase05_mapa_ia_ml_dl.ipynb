## Tres círculos, uno dentro de otro

Cuando alguien dice "usamos IA" y luego "entrenamos un modelo de deep learning" como si fueran sinónimos, está mezclando cosas de alcance muy distinto. Pensadlo como círculos concéntricos:

**Inteligencia Artificial (IA)** ⊃ **Machine Learning (ML)** ⊃ **Deep Learning (DL)**

Todo el Deep Learning es Machine Learning. Todo el Machine Learning es Inteligencia Artificial. Pero no al revés: hay muchísima IA que no es Machine Learning (por ejemplo, un sistema experto basado en reglas escritas a mano), y mucho Machine Learning que no llega a ser Deep Learning.

```python
# Una forma tonta pero efectiva de visualizar la relación de
# contención con conjuntos de Python: cada conjunto "más específico"
# es un subconjunto del anterior.
DL = {"redes neuronales convolucionales", "redes neuronales recurrentes", "transformers"}
ML = DL | {"regresion", "arboles de decision", "knn"}
IA = ML | {"reglas expertas", "busqueda", "logica difusa"}

print("¿DL es subconjunto de ML?", DL.issubset(ML))
print("¿ML es subconjunto de IA?", ML.issubset(IA))
print("Elementos que están en IA pero NO en ML:", IA - ML)
```

## Machine Learning: aprender de ejemplos, no de reglas

En programación tradicional, tú escribes las reglas y el ordenador las ejecuta. En Machine Learning, le das al algoritmo muchos ejemplos ya resueltos (datos de entrenamiento), y es el propio algoritmo el que descubre el patrón, generalizándolo para casos nuevos.

## Deep Learning: redes neuronales en capas

Un subconjunto especializado de ML que usa **redes neuronales artificiales** en capas para simular, de forma simplificada, cómo el cerebro toma decisiones. Diferencia clave: muchos algoritmos de ML "clásico" se estancan (hacen *plateau*) al darles más datos; el Deep Learning tiende a **mejorar** cuantos más datos recibe.

Una red neuronal es una colección de "neuronas" organizadas en capas que transforman la entrada paso a paso y se entrenan iterativamente. Cuando hay muchas capas, se dice que la red es "profunda" — de ahí el nombre. Esta técnica estuvo casi 15 años en desuso por su coste computacional, hasta que las **GPUs** (con cientos de núcleos en paralelo — una GPU equivale aproximadamente a 600 núcleos) la volvieron viable de nuevo.

## ¿IA y Ciencia de Datos son lo mismo?

No. No hay relación de subconjunto entre ellas. La **Ciencia de Datos** es el proceso de extraer conocimiento de grandes volúmenes de datos (mezcla de matemáticas, estadística, visualización, programación). La **IA** es todo lo que permite a las máquinas aprender a resolver problemas por sí mismas. Se cruzan constantemente: la ciencia de datos usa técnicas de IA (incluyendo ML y DL) como herramientas, pero no es un subconjunto de la IA ni al revés.

## Demo en vivo: reglas fijas vs. un modelo que aprende

Vamos a resolver el MISMO problema de dos formas distintas: clasificar flores del dataset `iris` (incluido en scikit-learn, sin descargas) usando solo dos medidas del pétalo.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

iris = load_iris()

# Nos quedamos con 2 especies (0 y 1) y 2 columnas: largo y ancho de pétalo
mascara = iris.target != 2
X = iris.data[mascara][:, [2, 3]]
y = iris.target[mascara]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

print("Ejemplos de entrenamiento:", len(X_train))
print("Ejemplos de prueba:", len(X_test))
```

```python
# --- Enfoque 1: reglas fijas, escritas a mano por nosotros ---
def clasificar_con_reglas(largo_petalo, ancho_petalo):
    if largo_petalo < 2.5:
        return 0  # setosa
    return 1  # versicolor

predicciones_reglas = [clasificar_con_reglas(l, a) for l, a in X_test]
aciertos_reglas = np.mean(predicciones_reglas == y_test)
print(f"Reglas fijas escritas a mano: {aciertos_reglas:.0%} de acierto")

# --- Enfoque 2: un modelo que APRENDE de los datos ---
modelo = KNeighborsClassifier(n_neighbors=3)
modelo.fit(X_train, y_train)  # aquí "aprende": memoriza los ejemplos y su geometría
aciertos_modelo = modelo.score(X_test, y_test)
print(f"Modelo entrenado (KNN): {aciertos_modelo:.0%} de acierto")
```

En este dataset tan "fácil" ambos enfoques pueden acertar mucho, porque las especies están bien separadas. Pero fijaos en la diferencia de proceso: nosotros **inventamos** el umbral `2.5` mirando los datos a ojo; el modelo KNN nunca vio ninguna regla — solo ejemplos ya etiquetados, y descubrió por sí mismo cómo separarlos comparando cada flor nueva con las más parecidas que ya conocía. En problemas reales, con docenas de variables, escribir reglas a mano deja de ser viable mucho antes que entrenar un modelo.
