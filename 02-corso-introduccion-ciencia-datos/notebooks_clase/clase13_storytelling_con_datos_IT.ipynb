## Perché conta questo?

Puoi passare settimane a fare un'analisi brillante e comunque fallire completamente se non sai comunicarla. I dati, da soli, non convincono nessuno — a convincere è la storia che costruisci intorno a loro. Ricorda il processo del capitolo 1: non finisce con "analizzare", finisce con "raccontare la storia" e "visualizzare".

## Quanto deve essere lungo un report?

Non c'è una risposta unica:
- Accademico: da 1.000 a 7.000 parole.
- Business/consulenza: da meno di 1.500 parole fino a centinaia di pagine.

Società come McKinsey e Deloitte hanno costruito buona parte della loro reputazione su report basati sull'analisi dei dati.

## Un esempio reale: il report di Deloitte

"United States Economic Forecast" (Deloitte University Press, dicembre 2014, 24 pagine):

- Apre con un'affermazione che cattura l'attenzione (l'economia è più sana di quanto si percepisca).
- Cita Voltaire per inquadrare il tono.
- Usa grafici a serie temporali del PIL (contrazione e ripresa della Grande Recessione).
- Presenta **quattro scenari futuri**, non un'unica previsione.
- Si collega a Thomas Piketty e al dibattito sulla disuguaglianza di reddito.
- Relega il dettaglio tecnico (4 tabelle di dati) a un'appendice.

**Argomentazione centrale:** questa narrazione persuade molto di più di quegli stessi 8 grafici e 4 tabelle presentati isolati, come diapositive senza filo conduttore.

## Pianifica il finale prima di iniziare

Consiglio chiave: **pianifica il report finale prima di iniziare l'analisi**, non dopo. Così, ogni decisione durante l'analisi (quali dati privilegiare, quali grafici preparare) è guidata dall'obiettivo di comunicazione.

Un buon report comunica: beneficio per il lettore, obiettivi chiari, importanza del tema, contesto, applicabilità pratica e direzioni future.

La struttura dipende dallo scopo e dalla lunghezza: i report brevi vanno dritti al sodo; quelli estesi costruiscono l'argomentazione progressivamente e seguono un formato più prescritto (copertina, indice, riassunto esecutivo, contenuto, ringraziamenti). Dato curioso: persino i candidati al dottorato a volte dimenticano la copertina.

## Demo dal vivo: da "mostrare dati" a "raccontare una storia"

Partiamo da una piccola tabella fatta a mano —una serie temporale con vari scenari, nello stile del report di Deloitte— e confrontiamo due modi di presentarla: grezza, e come una storia visiva con un'annotazione che segnala il punto chiave.

```python
import matplotlib.pyplot as plt

# Piccola tabella fatta a mano: "crescita economica proiettata" sotto tre scenari,
# ispirata al tipo di dato di un report di previsione economica.
anios = [2021, 2022, 2023, 2024, 2025]
escenario_pesimista  = [1.0, 0.5, 0.3, 0.8, 1.2]
escenario_base       = [1.0, 1.8, 2.1, 2.4, 2.6]
escenario_optimista  = [1.0, 2.5, 3.4, 4.0, 4.5]

# Passo 1: "mostrare dati" grezzi
print("Scenario base:", escenario_base)
```

```python
# Passo 2: "raccontare una storia" con lo stesso dato
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(anios, escenario_pesimista, label="Scenario pessimistico", color="firebrick", linestyle="--")
ax.plot(anios, escenario_base, label="Scenario base", color="steelblue", linewidth=2.5)
ax.plot(anios, escenario_optimista, label="Scenario ottimistico", color="seagreen", linestyle="--")

ax.set_title("Crescita economica proiettata sotto tre scenari")
ax.set_xlabel("Anno")
ax.set_ylabel("Crescita proiettata (%)")
ax.legend(loc="upper left")

# L'annotazione segnala esplicitamente il punto che vogliamo il pubblico ricordi
ax.annotate(
    "La ripresa accelera\na partire dal 2023",
    xy=(2023, escenario_base[2]),
    xytext=(2021.3, 3.6),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=10,
)

plt.tight_layout()
plt.show()
```

Confronta le due celle precedenti: la prima è informazione corretta ma senza focus — l'equivalente degli "8 grafici e 4 tabelle senza filo narrativo". La seconda organizza i tre scenari a colpo d'occhio, e la chiamata a `plt.annotate()` è ciò che trasforma il grafico in una storia: invece di aspettare che il pubblico individui da solo il cambiamento importante, glielo segnaliamo esplicitamente — proprio come Deloitte ha aperto il suo report con un'affermazione invece di lasciare che il lettore arrivasse da solo a quella conclusione 20 pagine dopo.

## Riepilogo della lezione

- I dati non persuadono da soli — persuade la narrazione intorno a loro.
- Un buon report combina: affermazione iniziale forte, contesto/riferimenti, visualizzazioni che riassumono la storia, riconoscimento dell'incertezza, e dettaglio tecnico relegato a un'appendice.
- Pianifica il report finale prima di analizzare, non dopo.
- Titolo, etichette degli assi e soprattutto `plt.annotate()` sono ciò che trasforma "mostrare un grafico" in "raccontare una storia con un grafico".
