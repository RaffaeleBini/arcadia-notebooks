## 1. Cos'è il cloud computing?

Il cloud computing è l'**erogazione di risorse di calcolo su richiesta** (reti, server, storage, applicazioni) attraverso internet, **pagando solo per ciò che usi**.

### Le 5 caratteristiche essenziali
1. **Self-service on demand**: accedi alle risorse senza interagire con una persona del fornitore.
2. **Accesso di rete ampio**: da cellulare, tablet, portatile o workstation.
3. **Pooling delle risorse**: il fornitore serve più clienti con risorse condivise dinamicamente (economie di scala), senza che tu sappia la posizione fisica.
4. **Elasticità rapida**: accedi a più risorse quando ti servono e le liberi quando non ti servono.
5. **Servizio misurato**: paghi per ciò che usi o prenoti, con reporting trasparente.

## 2. Modelli di implementazione e di servizio

### 3 modelli di implementazione
- **Pubblico**: infrastruttura di un fornitore, condivisa tra aziende, su internet aperto.
- **Privato**: infrastruttura esclusiva di un'organizzazione (on-premises o gestita esternamente).
- **Ibrido**: mix di entrambi che lavorano insieme.

### 3 modelli di servizio (in base allo strato della pila di calcolo)
- **IaaS** (Infrastructure as a Service): accedi a server/rete/storage senza gestirli.
- **PaaS** (Platform as a Service): accedi a una piattaforma hardware+software per sviluppare e pubblicare applicazioni.
- **SaaS** (Software as a Service): usi software/applicazioni ospitate e concesse in licenza in abbonamento — "software su richiesta".

## 3. Perché conta per un/una data scientist

### Demo dal vivo: simulare il "servizio misurato"

Non possiamo distribuire infrastruttura reale in un notebook — ma possiamo simulare la logica di fatturazione che rende possibile il servizio misurato: pagare solo per le ore di calcolo effettivamente usate.

```python
import pandas as pd

consumo_horas = {
    "lunes": 2.5, "martes": 0.0, "miercoles": 6.0,
    "jueves": 1.0, "viernes": 3.5, "sabado": 0.0, "domingo": 0.0,
}
precio_por_hora = 0.90

consumo = pd.DataFrame(list(consumo_horas.items()), columns=["dia", "horas_usadas"])
consumo["coste_usd"] = consumo["horas_usadas"] * precio_por_hora

print(consumo)
print(f"\nCosto totale della settimana (cloud pubblico, pagamento a uso): ${consumo['coste_usd'].sum():.2f}")

coste_servidor_propio = 24 * 7 * precio_por_hora
print(f"Costo se avessimo un server proprio acceso 24/7: ${coste_servidor_propio:.2f}")
```

La differenza tra i due costi è, in sostanza, il vantaggio economico dell'elasticità e del servizio misurato per carichi di lavoro intermittenti — molto tipici nella scienza dei dati (addestri un modello, finisci, e smetti di pagare).
