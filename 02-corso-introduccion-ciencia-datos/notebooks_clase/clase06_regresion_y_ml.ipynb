## El "caballo de batalla" del análisis estadístico

El nombre "regresión" viene de un estudio de **Sir Francis Galton (1886)** sobre la altura heredada entre padres e hijos: notó que las alturas extremas de los padres "regresaban" hacia la media de la población en cada generación — de ahí "regresión a la media".

Hoy, una **regresión** es una técnica estadística que mide la fuerza y dirección de la relación entre variables de entrada y una variable de salida. El resultado no es solo "sí, están relacionadas": es un **coeficiente** que te dice cuánto cambia la salida por cada unidad que cambia la entrada.

### Ejemplo real: precios de vivienda (tesis de Murtaza Haider, 1999, 500.000+ transacciones)

- Un baño adicional sube el precio **más** que un dormitorio adicional.
- Cercanía al metro: sube el precio. Cercanía a autopistas: lo baja.
- Cercanía a centros comerciales: efecto **no lineal** — muy cerca (<2.5 km) baja el precio; moderadamente cerca (2.5-5 km) lo sube.
- Más lejos del centro de la ciudad → precio más bajo.

Ninguno de estos hallazgos es intuitivo a simple vista — por eso hace falta correr los números.

## Más allá de la regresión

- **Sistemas de recomendación**: una de las aplicaciones de ML más usadas.
- **Clasificación**, **clústeres**, **cesta de la compra** (qué se compra junto).
- **Analítica predictiva**: árboles de decisión, Naive Bayes — cuidado con el compromiso **precisión/exhaustividad (recall)** y el riesgo de **sobreajuste (overfitting)**.
- **Fintech**: recomendaciones de inversión y **detección de fraude con tarjetas en tiempo real**, entrenada con transacciones históricas.

## Demo en vivo: entrenando una regresión lineal real

Usamos `load_diabetes()`, un dataset médico incluido en scikit-learn (100% local, sin descargas), para predecir la progresión de la enfermedad a partir del índice de masa corporal (BMI).

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target

variable_predictora = "bmi"
X_bmi = X[[variable_predictora]]

modelo = LinearRegression()
modelo.fit(X_bmi, y)

coeficiente = modelo.coef_[0]
interseccion = modelo.intercept_
print(f"Coeficiente de '{variable_predictora}': {coeficiente:.2f}")
print(f"Intersección: {interseccion:.2f}")
```

```python
plt.scatter(X_bmi, y, alpha=0.5, label="Pacientes reales")

predicciones = modelo.predict(X_bmi)
orden = np.argsort(X_bmi[variable_predictora].values)
plt.plot(
    X_bmi[variable_predictora].values[orden],
    predicciones[orden],
    color="red", linewidth=2, label="Línea de regresión"
)
plt.xlabel("Índice de masa corporal (normalizado)")
plt.ylabel("Progresión de la enfermedad")
plt.legend()
plt.title("Regresión lineal: BMI vs. progresión de la diabetes")
plt.show()
```

El coeficiente positivo nos dice que, según este modelo, a mayor índice de masa corporal, mayor progresión de la enfermedad predicha. La línea roja es la relación que aprendió el modelo — no pasa por todos los puntos (hay muchos más factores en juego que solo el BMI), pero captura la tendencia general. Es la misma idea de Galton con las alturas, aplicada aquí a datos médicos con la ayuda de `scikit-learn`.
