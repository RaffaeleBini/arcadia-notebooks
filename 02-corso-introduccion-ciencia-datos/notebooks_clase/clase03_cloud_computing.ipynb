## 1. ¿Qué es la computación en la nube?

La computación en la nube (cloud computing) es la **entrega de recursos de cómputo bajo demanda** (redes, servidores, almacenamiento, aplicaciones) a través de internet, **pagando solo por lo que usas**.

### Las 5 características esenciales
1. **Autoservicio bajo demanda**: accedes a recursos sin interactuar con un humano del proveedor.
2. **Acceso de red amplio**: desde móvil, tablet, portátil o estación de trabajo.
3. **Pooling de recursos**: el proveedor sirve a múltiples clientes con recursos compartidos dinámicamente (economías de escala), sin que sepas la ubicación física.
4. **Elasticidad rápida**: accedes a más recursos cuando los necesitas y los liberas cuando no.
5. **Servicio medido**: pagas por lo que usas o reservas, con reporting transparente.

## 2. Modelos de despliegue y de servicio

### 3 modelos de despliegue
- **Público**: infraestructura de un proveedor, compartida entre empresas, sobre internet abierto.
- **Privado**: infraestructura exclusiva de una organización (on-premises o gestionada externamente).
- **Híbrido**: mezcla de ambos trabajando juntos.

### 3 modelos de servicio (según la capa de la pila de cómputo)
- **IaaS** (Infraestructura como Servicio): accedes a servidores/red/almacenamiento sin gestionarlos.
- **PaaS** (Plataforma como Servicio): accedes a una plataforma de hardware+software para desarrollar y desplegar aplicaciones.
- **SaaS** (Software como Servicio): usas software/aplicaciones alojadas y licenciadas por suscripción — "software bajo demanda".

### Demo en vivo: simulando el "servicio medido"

No podemos desplegar infraestructura real en un notebook — pero sí podemos simular la lógica de facturación que hace posible el servicio medido: pagar solo por las horas de cómputo efectivamente usadas.

```python
import pandas as pd

consumo_horas = {
    "lunes": 2.5, "martes": 0.0, "miercoles": 6.0,
    "jueves": 1.0, "viernes": 3.5, "sabado": 0.0, "domingo": 0.0,
}
precio_por_hora = 0.90

consumo = pd.DataFrame(list(consumo_horas.items()), columns=["dia", "horas_usadas"])
consumo["coste_usd"] = consumo["horas_usadas"] * precio_por_hora

print(consumo)
print(f"\nCoste total de la semana (nube pública, pago por uso): ${consumo['coste_usd'].sum():.2f}")

coste_servidor_propio = 24 * 7 * precio_por_hora
print(f"Coste si tuviéramos un servidor propio encendido 24/7: ${coste_servidor_propio:.2f}")
```

La diferencia entre ambos costes es, en esencia, la ventaja económica de la elasticidad y el servicio medido para cargas de trabajo intermitentes — muy típicas en ciencia de datos (entrenas un modelo, terminas, y dejas de pagar).
