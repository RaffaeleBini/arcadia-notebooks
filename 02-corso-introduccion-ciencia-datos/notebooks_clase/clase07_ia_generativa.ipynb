## 1. De analizar datos a crear datos

Todo lo que hemos visto hasta ahora —regresión, clasificación, machine learning en general— parte de datos que **ya existen** y busca patrones en ellos. La IA generativa da la vuelta al problema: ¿puede una máquina **crear** datos nuevos que no existían, pero que sean plausibles?

Ejemplos que seguramente ya conoces sin llamarlos por su nombre técnico:
- Un chatbot que te escribe un párrafo coherente.
- Una imagen generada a partir de una descripción de texto.
- Música o código generados automáticamente.

Todo esto es **IA generativa**: un subconjunto de la IA centrado en producir datos nuevos que imitan creaciones humanas.

## 2. Cómo funciona (a grandes rasgos)

Se apoya en modelos de deep learning entrenados para generar, no solo para predecir:

- **GAN (Redes Generativas Antagónicas)**: dos redes que compiten — una genera datos falsos, otra intenta detectarlos. Compitiendo, ambas mejoran.
- **VAE (Autocodificadores Variacionales)**: aprenden a comprimir datos reales y reconstruirlos, generando variaciones nuevas y plausibles.

Ambos enfoques aprenden la **distribución** de los datos reales y generan instancias nuevas que la respetan, sin copiar ejemplos existentes.

## 3. El uso que más nos interesa: datos sintéticos

Entrenar y probar un modelo requiere muchos datos, y a veces no hay suficientes observaciones reales (son caras, escasas, o están protegidas por privacidad).

Los data scientists usan IA generativa para crear **datos sintéticos**: datos artificiales con propiedades similares a las reales (misma distribución, mismos agrupamientos), que se combinan con los datos reales para entrenar y probar modelos con más robustez.

No hace falta una GAN para entender la idea de fondo. Si conoces la media y la desviación estándar de una variable real, puedes generar observaciones nuevas que sigan esa misma distribución. Vamos a hacerlo en vivo.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes

# Semilla fija para que el resultado sea reproducible en clase
np.random.seed(42)

# Dataset real, incluido en scikit-learn (no se descarga nada de internet)
diabetes = load_diabetes()
print("Forma del dataset:", diabetes.data.shape)
print("Nombres de columnas:", diabetes.feature_names)
```

```python
# Nos quedamos con una columna real: el índice de masa corporal (bmi)
bmi_real = diabetes.data[:, 2]

# Paso 1: medimos la distribución real
media_real = bmi_real.mean()
desviacion_real = bmi_real.std()

print(f"Media real:      {media_real:.4f}")
print(f"Desviación real:  {desviacion_real:.4f}")
print(f"Nº de observaciones reales: {len(bmi_real)}")
```

```python
# Paso 2: generamos datos sintéticos con la misma distribución
n_sinteticos = 1000
bmi_sintetico = np.random.normal(media_real, desviacion_real, n_sinteticos)

print(f"Media sintética:      {bmi_sintetico.mean():.4f}")
print(f"Desviación sintética: {bmi_sintetico.std():.4f}")
print(f"Nº de observaciones sintéticas: {len(bmi_sintetico)}")
```

```python
# Paso 3: comparamos visualmente ambas distribuciones
plt.figure(figsize=(8, 5))
plt.hist(bmi_real, bins=25, alpha=0.6, density=True, label="Datos reales")
plt.hist(bmi_sintetico, bins=25, alpha=0.6, density=True, label="Datos sintéticos")
plt.title("BMI real vs. BMI sintético")
plt.xlabel("Índice de masa corporal (normalizado)")
plt.ylabel("Densidad")
plt.legend()
plt.show()
```

Fíjate en que las dos campanas se superponen casi por completo: eso es evidencia visual de que los datos sintéticos "se comportan" como los reales, aunque ninguna de esas 1.000 observaciones corresponda a un paciente real.

Esta es, en miniatura, la misma lógica que usa un data scientist cuando necesita ampliar un dataset demasiado pequeño para entrenar un modelo: aprender la distribución de lo real y generar más ejemplos que la respeten.

## 4. Otras aplicaciones de la IA generativa

- Generación de texto tipo humano (GPT-3) para chatbots y creación de contenido.
- Síntesis de imágenes médicas para formación de profesionales de salud.
- Arte generado, entornos y niveles de videojuegos, diseño de moda.
- **Automatización de código**: generar y probar automáticamente código de modelos analíticos.
- **Generación de insights e informes** que se actualizan según evolucionan los datos (ejemplo: IBM Cognos Analytics, con un asistente en lenguaje natural).

## Resumen de la clase

- La IA generativa **crea** datos nuevos en vez de solo analizar los existentes.
- Se apoya en modelos como GAN y VAE, que aprenden la distribución de los datos reales.
- Uno de sus usos más prácticos para un data scientist: generar **datos sintéticos** cuando faltan datos reales suficientes.
- `np.random.normal(media, desviación, tamaño)` es una forma sencilla de generar datos sintéticos que sigan la distribución de una variable real.
