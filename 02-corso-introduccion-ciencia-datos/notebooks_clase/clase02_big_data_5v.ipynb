## 1. ¿Por qué hablamos de "Big Data"?

Cada vez que usas el móvil, un reloj inteligente o una app de streaming dejas un rastro digital. De esa acumulación masiva y constante de datos nace el término **Big Data**.

Definición de Ernst & Young: *"Big Data se refiere a los volúmenes dinámicos, grandes y dispares de datos creados por personas, herramientas y máquinas. Requiere tecnología nueva, innovadora y escalable para recolectar, almacenar y procesar analíticamente esa cantidad de datos con el fin de obtener insights de negocio en tiempo real relacionados con consumidores, riesgo, ganancias, rendimiento, productividad y valor para los accionistas"*.

No existe una única definición universal — pero casi todas coinciden en 5 elementos comunes: **las 5V**.

## 2. Las 5V

| V | Qué mide | Ejemplo |
|---|---|---|
| **Volumen** | La escala de los datos | ~7.000 millones de personas generan ~2,5 trillones de bytes/día |
| **Velocidad** | La rapidez de acumulación | Cada 60 segundos se suben horas de vídeo a YouTube |
| **Variedad** | La diversidad de formatos/fuentes | Texto, imágenes, vídeo, wearables, IoT |
| **Veracidad** | La calidad y fiabilidad | ~80% de los datos es no estructurado |
| **Valor** | La capacidad de convertir datos en algo útil | Económico, médico, social... |

Definición alternativa y práctica: Big Data es aquello que ya **no puedes manejar con sistemas de bases de datos tradicionales**, por su volumen y velocidad combinados. Origen histórico: Google (Larry Page y Sergey Brin) tuvo que inventar tecnología nueva para el algoritmo PageRank y para almacenar toda la web — de ahí nace lo que después inspiraría a Hadoop (próxima clase).

### Demo en vivo: midiendo Volumen, Variedad y Veracidad

Simulamos 1000 eventos de sensores de una ciudad inteligente y medimos tres de las 5V directamente sobre el DataFrame.

```python
import pandas as pd
import numpy as np

np.random.seed(3)
n_eventos = 1000

tipos_sensor = np.random.choice(["trafico", "clima", "wearable", "camara"], size=n_eventos)
medida = np.random.normal(loc=50, scale=15, size=n_eventos)

# Simulamos algo de "ruido" real: valores faltantes y erróneos
idx_faltantes = np.random.choice(n_eventos, size=int(n_eventos * 0.05), replace=False)
medida[idx_faltantes] = np.nan
idx_erroneos = np.random.choice(n_eventos, size=int(n_eventos * 0.02), replace=False)
medida[idx_erroneos] = -999

eventos = pd.DataFrame({"tipo_sensor": tipos_sensor, "medida": medida})

print("VOLUMEN (num. de eventos):", len(eventos))
print("\nVARIEDAD (tipos de sensor):")
print(eventos["tipo_sensor"].value_counts())

faltantes = eventos["medida"].isna().sum()
erroneos = (eventos["medida"] == -999).sum()
utilizables = len(eventos) - faltantes - erroneos
print(f"\nVERACIDAD: {utilizables}/{len(eventos)} registros utilizables ({utilizables/len(eventos):.1%})")
```

Fíjate cómo el Volumen (`len`), la Variedad (`value_counts`) y la Veracidad (proporción de datos limpios) se pueden medir con herramientas de pandas que ya conoces — pero a escala real (millones o miles de millones de filas) estas mismas operaciones necesitan computación distribuida, tema de la próxima clase.

## 3. ¿Por qué no bastan las herramientas tradicionales?

Las herramientas convencionales de análisis (piensa en abrir un archivo gigante en una hoja de cálculo) **no son viables** a escala de Big Data. Hacen falta herramientas de **computación distribuida** — repartir el trabajo entre muchos ordenadores a la vez. De eso trata el ecosistema Hadoop, Hive y Spark, que veremos más adelante en el curso.

## Resumen de la clase

- Big Data no tiene una única definición, pero casi todas comparten las 5V: Volumen, Velocidad, Variedad, Veracidad y Valor.
- Una definición práctica: Big Data es lo que ya no cabe en un sistema de bases de datos tradicional.
- El 80% de los datos generados hoy es no estructurado, lo que complica garantizar Veracidad.
- Las herramientas convencionales no escalan a este volumen; hacen falta herramientas de computación distribuida.
