## Il "cavallo di battaglia" dell'analisi statistica

Il nome "regressione" viene da uno studio di **Sir Francis Galton (1886)** sull'altezza ereditata tra genitori e figli: notò che le altezze estreme dei genitori "regredivano" verso la media della popolazione a ogni generazione — da qui "regressione verso la media".

Oggi, una **regressione** è una tecnica statistica che misura la forza e la direzione della relazione tra variabili in ingresso e una variabile in uscita. Il risultato non è solo "sì, sono collegate": è un **coefficiente** che ti dice quanto cambia l'uscita per ogni unità che cambia l'ingresso.

### Esempio reale: prezzi delle abitazioni (tesi di Murtaza Haider, 1999, 500.000+ transazioni)

- Un bagno aggiuntivo fa salire il prezzo **più** di una camera da letto aggiuntiva.
- Vicinanza alla metro: fa salire il prezzo. Vicinanza alle autostrade: lo fa scendere.
- Vicinanza ai centri commerciali: effetto **non lineare** — molto vicino (<2,5 km) fa scendere il prezzo; moderatamente vicino (2,5-5 km) lo fa salire.
- Più lontano dal centro città → prezzo più basso.

Nessuna di queste scoperte è intuitiva a prima vista — per questo bisogna fare i calcoli.

## Oltre la regressione

- **Sistemi di raccomandazione**: una delle applicazioni di ML più usate.
- **Classificazione**, **cluster**, **paniere della spesa** (cosa si compra insieme).
- **Analisi predittiva**: alberi decisionali, Naive Bayes — attenzione al compromesso **precisione/esaustività (recall)** e al rischio di **sovradattamento (overfitting)**.
- **Fintech**: raccomandazioni di investimento e **rilevamento di frodi con carte in tempo reale**, addestrato con transazioni storiche.

## Demo dal vivo: addestrare una vera regressione lineare

Usiamo `load_diabetes()`, un dataset medico incluso in scikit-learn (100% locale, senza download), per predire la progressione della malattia a partire dall'indice di massa corporea (BMI).

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target

variable_predictora = "bmi"
X_bmi = X[[variable_predictora]]

modelo = LinearRegression()
modelo.fit(X_bmi, y)

coeficiente = modelo.coef_[0]
interseccion = modelo.intercept_
print(f"Coefficiente di '{variable_predictora}': {coeficiente:.2f}")
print(f"Intercetta: {interseccion:.2f}")
```

```python
plt.scatter(X_bmi, y, alpha=0.5, label="Pazienti reali")

predicciones = modelo.predict(X_bmi)
orden = np.argsort(X_bmi[variable_predictora].values)
plt.plot(
    X_bmi[variable_predictora].values[orden],
    predicciones[orden],
    color="red", linewidth=2, label="Retta di regressione"
)
plt.xlabel("Indice di massa corporea (normalizzato)")
plt.ylabel("Progressione della malattia")
plt.legend()
plt.title("Regressione lineare: BMI vs. progressione del diabete")
plt.show()
```

Il coefficiente positivo ci dice che, secondo questo modello, a un indice di massa corporea maggiore corrisponde una progressione della malattia predetta maggiore. La retta rossa è la relazione appresa dal modello — non passa per tutti i punti (ci sono molti più fattori in gioco oltre al solo BMI), ma cattura la tendenza generale. È la stessa idea di Galton con le altezze, applicata qui a dati medici con l'aiuto di `scikit-learn`.
