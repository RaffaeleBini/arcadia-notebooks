```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes

np.random.seed(7)

diabetes = load_diabetes()
bp_real = diabetes.data[:, 3]

# Media e deviazione standard reali
media_real = bp_real.mean()
desviacion_real = bp_real.std()

print(f"Media reale: {media_real:.4f}")
print(f"Deviazione reale: {desviacion_real:.4f}")

# Generiamo 500 osservazioni sintetiche con la stessa distribuzione
n_sinteticos = 500
bp_sintetico = np.random.normal(media_real, desviacion_real, n_sinteticos)

print(f"Media sintetica: {bp_sintetico.mean():.4f}")
print(f"Deviazione sintetica: {bp_sintetico.std():.4f}")

# Confronto visivo
plt.figure(figsize=(8, 5))
plt.hist(bp_real, bins=25, alpha=0.6, density=True, label="Dati reali")
plt.hist(bp_sintetico, bins=25, alpha=0.6, density=True, label="Dati sintetici")
plt.title("Pressione sanguigna: reale vs. sintetica")
plt.xlabel("Pressione sanguigna media (normalizzata)")
plt.ylabel("Densità")
plt.legend()
plt.show()
```

```python
diabetes = load_diabetes()
bmi_real = diabetes.data[:, 2]
media_real = bmi_real.mean()
desviacion_real = bmi_real.std()

tamanos = [20, 200, 2000]

np.random.seed(0)
muestras_sinteticas = []
for tamano in tamanos:
    muestra = np.random.normal(media_real, desviacion_real, tamano)
    muestras_sinteticas.append(muestra)

for tamano, muestra in zip(tamanos, muestras_sinteticas):
    print(f"n={tamano:5d}  media={muestra.mean():.4f}  deviazione={muestra.std():.4f}"
          f"   (reale: media={media_real:.4f}, deviazione={desviacion_real:.4f})")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, tamano, muestra in zip(axes, tamanos, muestras_sinteticas):
    ax.hist(bmi_real, bins=20, alpha=0.6, density=True, label="Reale")
    ax.hist(muestra, bins=20, alpha=0.6, density=True, label="Sintetico")
    ax.set_title(f"n sintetico = {tamano}")
    ax.legend()
plt.tight_layout()
plt.show()

# Conclusione attesa: con n=20 il campione sintetico è "rumoroso" e la sua
# media/deviazione si allontanano di più da quelle reali; con n=2000 si avvicina molto
# di più alla distribuzione reale. Quanti più dati sintetici generiamo, più
# fedelmente si avvicinano alla distribuzione che abbiamo imparato dai dati reali.
```
