## Ejercicio 1 — Cuenta la historia de una campaña de marketing

Tienes el número de visitas mensuales a la web de una tienda a lo largo de un año. La empresa lanzó una campaña de publicidad en julio y quiere saber si se nota en los datos, contado como una historia y no solo como una tabla.

**Pistas:**
- Usa `ax.plot()` para el gráfico de líneas.
- Añade título (`set_title`) y etiquetas de ejes (`set_xlabel`, `set_ylabel`).
- Usa `ax.annotate()` para señalar el mes de la campaña (julio) y explicar qué pasó ahí.

````python
import matplotlib.pyplot as plt

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun", "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]
visitas = [1200, 1150, 1300, 1250, 1400, 1380, 2600, 2400, 2100, 2050, 2000, 1950]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(meses, visitas, color="steelblue", linewidth=2)

# TODO: añade un título con ax.set_title() que resuma la historia de este gráfico
# TODO: añade etiquetas de ejes con ax.set_xlabel() y ax.set_ylabel()
# TODO: añade una anotación con ax.annotate() que señale el mes "Jul" (índice 6 en la
#       lista "meses") y explique que ahí empezó la campaña de publicidad

plt.tight_layout()
plt.show()
````

```python
import matplotlib.pyplot as plt

meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun", "Jul", "Ago", "Sep", "Oct", "Nov", "Dic"]
visitas = [1200, 1150, 1300, 1250, 1400, 1380, 2600, 2400, 2100, 2050, 2000, 1950]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(meses, visitas, color="steelblue", linewidth=2)

ax.set_title("Visitas mensuales a la web: efecto de la campaña de julio")
ax.set_xlabel("Mes")
ax.set_ylabel("Visitas")

ax.annotate(
    "Lanzamiento de la campaña\nde publicidad",
    xy=(6, visitas[6]),
    xytext=(3, 2400),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=10,
)

plt.tight_layout()
plt.show()
```

## Ejercicio 2 — Elige el escenario que resaltas

Tienes tres escenarios de ventas proyectadas para el próximo año (pesimista, base, optimista), igual que en la demo de clase, pero esta vez la anotación debe resaltar el escenario **pesimista** en el último trimestre, porque es el que más preocupa al equipo directivo.

**Pistas:**
- El código base ya dibuja las tres líneas — no hace falta que lo toques.
- Solo te falta añadir la llamada a `ax.annotate()` sobre `escenario_pesimista`.

````python
import matplotlib.pyplot as plt

trimestres = ["T1", "T2", "T3", "T4"]
escenario_pesimista = [100, 95, 80, 60]
escenario_base = [100, 105, 110, 115]
escenario_optimista = [100, 115, 130, 150]

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(trimestres, escenario_pesimista, label="Pesimista", color="firebrick", linestyle="--")
ax.plot(trimestres, escenario_base, label="Base", color="steelblue", linewidth=2.5)
ax.plot(trimestres, escenario_optimista, label="Optimista", color="seagreen", linestyle="--")
ax.set_title("Ventas proyectadas por trimestre")
ax.set_xlabel("Trimestre")
ax.set_ylabel("Ventas (índice, T1 = 100)")
ax.legend()

# TODO: añade una anotación con ax.annotate() que resalte el valor de
#       escenario_pesimista en T4 (índice 3), avisando de la caída esperada

plt.tight_layout()
plt.show()
````

```python
import matplotlib.pyplot as plt

trimestres = ["T1", "T2", "T3", "T4"]
escenario_pesimista = [100, 95, 80, 60]
escenario_base = [100, 105, 110, 115]
escenario_optimista = [100, 115, 130, 150]

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(trimestres, escenario_pesimista, label="Pesimista", color="firebrick", linestyle="--")
ax.plot(trimestres, escenario_base, label="Base", color="steelblue", linewidth=2.5)
ax.plot(trimestres, escenario_optimista, label="Optimista", color="seagreen", linestyle="--")
ax.set_title("Ventas proyectadas por trimestre")
ax.set_xlabel("Trimestre")
ax.set_ylabel("Ventas (índice, T1 = 100)")
ax.legend()

ax.annotate(
    "Riesgo: caída del 40%\nen el escenario pesimista",
    xy=(3, escenario_pesimista[3]),
    xytext=(1, 60),
    arrowprops=dict(arrowstyle="->", color="firebrick"),
    fontsize=10,
    color="firebrick",
)

plt.tight_layout()
plt.show()
```
