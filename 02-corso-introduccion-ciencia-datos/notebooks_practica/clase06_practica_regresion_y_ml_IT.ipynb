## Esercizio 1: scegli un'altra variabile e confronta l'errore

**Il tuo compito**:
1. Scegli una variabile predittiva diversa da `bmi` (controlla `diabetes.feature_names` per vedere le opzioni: `age`, `sex`, `bp`, `s1`...`s6`).
2. Addestra un secondo modello `LinearRegression` usando solo quella variabile.
3. Calcola l'errore quadratico medio (`mean_squared_error`) di ENTRAMBI i modelli sugli stessi dati e confrontali: quale sbaglia meno?

Suggerimento: un `mean_squared_error` più basso significa che le predizioni del modello sono, in media, più vicine ai valori reali.

````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import numpy as np

diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target

print("Variabili disponibili:", list(diabetes.feature_names))

# --- Modello 1: bmi (come in lezione, già risolto) ---
X_bmi = X[["bmi"]]
modelo_bmi = LinearRegression()
modelo_bmi.fit(X_bmi, y)
pred_bmi = modelo_bmi.predict(X_bmi)
error_bmi = mean_squared_error(y, pred_bmi)
print(f"MSE con 'bmi': {error_bmi:.2f}")

# --- Modello 2: la tua scelta ---
# TODO: scegli una variabile da diabetes.feature_names diversa da "bmi"
otra_variable = "bmi"  # <- cambiala con un'altra, ad esempio "bp" o "s5"

X_otra = X[[otra_variable]]

# TODO: crea un LinearRegression, addestralo con X_otra/y, e salva
# le predizioni in `pred_otra`
modelo_otra = None
pred_otra = None

# TODO: calcola l'errore quadratico medio di `pred_otra` rispetto a `y`
error_otra = None

print(f"MSE con '{otra_variable}': {error_otra}")
````
