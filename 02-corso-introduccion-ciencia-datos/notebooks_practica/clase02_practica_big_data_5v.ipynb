## Ejercicio: simulando un flujo de eventos en tiempo real

Vas a generar con NumPy un dataset sintético que simula eventos llegando a un sistema con **timestamps crecientes** — como si fueran clics en una web, transacciones de tarjeta o mensajes de sensores IoT.

Con ese dataset, vas a calcular:
1. **Velocidad**: cuántos eventos llegan por minuto en promedio.
2. **Volumen**: el número total de eventos y de bytes simulados generados.
3. **Variedad**: cuántos tipos distintos de evento aparecen y su distribución de tamaño (dos columnas de tipos de datos distintos: categórica y numérica).

**Pista**: para la Velocidad, una vez tengas una columna de timestamps, la diferencia entre el primer y el último timestamp (en minutos) te da la ventana de tiempo total; divide el número de eventos entre esa ventana.

````python
import pandas as pd
import numpy as np

np.random.seed(21)

n_eventos = 5000

# Generamos timestamps crecientes: cada evento llega entre 0 y 3 segundos después del anterior
# (simulando Velocidad: eventos casi en tiempo real, con ritmo irregular)
segundos_entre_eventos = np.random.exponential(scale=0.8, size=n_eventos)
segundos_acumulados = np.cumsum(segundos_entre_eventos)
timestamp_inicio = pd.Timestamp("2026-01-01 00:00:00")
timestamps = timestamp_inicio + pd.to_timedelta(segundos_acumulados, unit="s")

# Variedad: tipo de evento (columna categórica) y tamaño en bytes (columna numérica)
tipos_evento = np.random.choice(
    ["clic", "compra", "login", "error"], size=n_eventos, p=[0.6, 0.15, 0.2, 0.05]
)
tamano_bytes = np.random.randint(200, 5000, size=n_eventos)

eventos = pd.DataFrame({
    "timestamp": timestamps,
    "tipo_evento": tipos_evento,
    "tamano_bytes": tamano_bytes,
})

print(eventos.head())
print("\nTotal de eventos generados:", len(eventos))

# TODO 1 (VOLUMEN): calcula el total de bytes generados por todos los eventos juntos
# (suma la columna 'tamano_bytes')
volumen_total_bytes = None

# TODO 2 (VELOCIDAD): calcula la ventana de tiempo total en minutos
# (resta el primer timestamp al último, y convierte el resultado a minutos con .total_seconds() / 60)
ventana_minutos = None

# TODO 3 (VELOCIDAD): calcula cuántos eventos llegan por minuto en promedio
# (número total de eventos dividido entre ventana_minutos)
eventos_por_minuto = None

# TODO 4 (VARIEDAD): calcula cuántos tipos distintos de evento hay y su distribución (value_counts)
variedad_tipos = None

print("\nVolumen total (bytes):", volumen_total_bytes)
print("Ventana de tiempo (minutos):", ventana_minutos)
print("Eventos por minuto (Velocidad):", eventos_por_minuto)
print("\nVariedad de tipos de evento:")
print(variedad_tipos)
````
