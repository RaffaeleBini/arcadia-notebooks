## Esercizio 1 — Racconta la storia di una campagna di marketing

Hai il numero di visite mensili al sito di un negozio nel corso di un anno. L'azienda ha lanciato una campagna pubblicitaria a luglio e vuole sapere se si nota nei dati, raccontato come una storia e non solo come una tabella.

**Suggerimenti:**
- Usa `ax.plot()` per il grafico a linee.

import matplotlib.pyplot as plt

## Esercizio 2 — Scegli lo scenario da evidenziare

Hai tre scenari di vendite proiettate per il prossimo anno (pessimistico, base, ottimistico), come nella demo di lezione, ma questa volta l'annotazione deve evidenziare lo scenario **pessimistico** nell'ultimo trimestre, perché è quello che preoccupa di più il team dirigenziale.

**Suggerimenti:**
- Il codice di base disegna già le tre linee — non serve toccarlo.
- Ti manca solo aggiungere la chiamata a `ax.annotate()` su `escenario_pesimista`.

````python
import matplotlib.pyplot as plt

trimestres = ["T1", "T2", "T3", "T4"]
escenario_pesimista = [100, 95, 80, 60]
escenario_base = [100, 105, 110, 115]
escenario_optimista = [100, 115, 130, 150]

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(trimestres, escenario_pesimista, label="Pessimistico", color="firebrick", linestyle="--")
ax.plot(trimestres, escenario_base, label="Base", color="steelblue", linewidth=2.5)
ax.plot(trimestres, escenario_optimista, label="Ottimistico", color="seagreen", linestyle="--")
ax.set_title("Vendite proiettate per trimestre")
ax.set_xlabel("Trimestre")
ax.set_ylabel("Vendite (indice, T1 = 100)")
ax.legend()

# TODO: aggiungi un'annotazione con ax.annotate() che evidenzi il valore di
#       escenario_pesimista in T4 (indice 3), avvisando del calo previsto

plt.tight_layout()
plt.show()
````

import matplotlib.pyplot as plt
