```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes

np.random.seed(7)

diabetes = load_diabetes()
bp_real = diabetes.data[:, 3]

# Media y desviación estándar reales
media_real = bp_real.mean()
desviacion_real = bp_real.std()

print(f"Media real: {media_real:.4f}")
print(f"Desviación real: {desviacion_real:.4f}")

# Generamos 500 observaciones sintéticas con la misma distribución
n_sinteticos = 500
bp_sintetico = np.random.normal(media_real, desviacion_real, n_sinteticos)

print(f"Media sintética: {bp_sintetico.mean():.4f}")
print(f"Desviación sintética: {bp_sintetico.std():.4f}")

# Comparación visual
plt.figure(figsize=(8, 5))
plt.hist(bp_real, bins=25, alpha=0.6, density=True, label="Datos reales")
plt.hist(bp_sintetico, bins=25, alpha=0.6, density=True, label="Datos sintéticos")
plt.title("Presión sanguínea: real vs. sintética")
plt.xlabel("Presión sanguínea media (normalizada)")
plt.ylabel("Densidad")
plt.legend()
plt.show()
```

```python
diabetes = load_diabetes()
bmi_real = diabetes.data[:, 2]
media_real = bmi_real.mean()
desviacion_real = bmi_real.std()

tamanos = [20, 200, 2000]

np.random.seed(0)
muestras_sinteticas = []
for tamano in tamanos:
    muestra = np.random.normal(media_real, desviacion_real, tamano)
    muestras_sinteticas.append(muestra)

for tamano, muestra in zip(tamanos, muestras_sinteticas):
    print(f"n={tamano:5d}  media={muestra.mean():.4f}  desviacion={muestra.std():.4f}"
          f"   (real: media={media_real:.4f}, desviacion={desviacion_real:.4f})")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, tamano, muestra in zip(axes, tamanos, muestras_sinteticas):
    ax.hist(bmi_real, bins=20, alpha=0.6, density=True, label="Real")
    ax.hist(muestra, bins=20, alpha=0.6, density=True, label="Sintético")
    ax.set_title(f"n sintético = {tamano}")
    ax.legend()
plt.tight_layout()
plt.show()

# Conclusión esperada: con n=20 la muestra sintética es "ruidosa" y su
# media/desviación se alejan más de las reales; con n=2000 se acerca mucho
# más a la distribución real. Cuantos más datos sintéticos generamos, más
# fielmente se aproximan a la distribución que aprendimos de los datos reales.
```
