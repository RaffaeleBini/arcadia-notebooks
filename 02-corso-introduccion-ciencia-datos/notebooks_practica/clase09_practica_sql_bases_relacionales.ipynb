- Puedes reutilizar la misma estructura de `JOIN` que viste en la clase de teoría.

````python
import sqlite3
import pandas as pd

# Andamiaje: recreamos las mismas tablas de la clase de teoría.
conexion = sqlite3.connect(':memory:')
cursor = conexion.cursor()

cursor.execute('''
    CREATE TABLE clientes (
        id_cliente INTEGER PRIMARY KEY,
        nombre TEXT NOT NULL,
        ciudad TEXT
    )
''')
cursor.execute('''
    CREATE TABLE transacciones (
        id_transaccion INTEGER PRIMARY KEY,
        id_cliente INTEGER,
        importe REAL NOT NULL,
        FOREIGN KEY (id_cliente) REFERENCES clientes (id_cliente)
    )
''')

clientes_nuevos = [
    (1, 'Ana Torres', 'Madrid'),
    (2, 'Luis Peña', 'Bogotá'),
    (3, 'Carla Ruiz', 'Madrid'),
    (4, 'Diego Soto', 'CDMX'),
]
cursor.executemany(
    'INSERT INTO clientes (id_cliente, nombre, ciudad) VALUES (?, ?, ?)',
    clientes_nuevos
)

transacciones_nuevas = [
    (101, 1, 59.90),
    (102, 1, 120.00),
    (103, 2, 34.50),
    (104, 3, 89.99),
    (105, 3, 15.00),
    (106, 4, 200.00),
]
cursor.executemany(
    'INSERT INTO transacciones (id_transaccion, id_cliente, importe) VALUES (?, ?, ?)',
    transacciones_nuevas
)
conexion.commit()

print("Tablas listas. Ahora escribe tus consultas en las celdas de abajo.")
````

    JOIN clientes ON transacciones.id_cliente = clientes.id_cliente

    ORDER BY gasto_total DESC
