## Esercizio: simulare un flusso di eventi in tempo reale

Genererai con NumPy un dataset sintetico che simula eventi in arrivo a un sistema con **timestamp crescenti** — come se fossero clic su un sito web, transazioni con carta o messaggi di sensori IoT.

Con quel dataset, calcolerai:
1. **Velocità**: quanti eventi arrivano al minuto in media.
2. **Volume**: il numero totale di eventi e di byte simulati generati.
3. **Varietà**: quanti tipi diversi di evento compaiono e la loro distribuzione dimensionale (due colonne di tipi di dato diversi: categorica e numerica).

**Suggerimento**: per la Velocità, una volta che hai una colonna di timestamp, la differenza tra il primo e l'ultimo timestamp (in minuti) ti dà la finestra di tempo totale; dividi il numero di eventi per quella finestra.

````python
import pandas as pd
import numpy as np

np.random.seed(21)

n_eventos = 5000

# Generiamo timestamp crescenti: ogni evento arriva tra 0 e 3 secondi dopo il precedente
# (simulando la Velocità: eventi quasi in tempo reale, con ritmo irregolare)
segundos_entre_eventos = np.random.exponential(scale=0.8, size=n_eventos)
segundos_acumulados = np.cumsum(segundos_entre_eventos)
timestamp_inicio = pd.Timestamp("2026-01-01 00:00:00")
timestamps = timestamp_inicio + pd.to_timedelta(segundos_acumulados, unit="s")

# Varietà: tipo di evento (colonna categorica) e dimensione in byte (colonna numerica)
tipos_evento = np.random.choice(
    ["clic", "compra", "login", "error"], size=n_eventos, p=[0.6, 0.15, 0.2, 0.05]
)
tamano_bytes = np.random.randint(200, 5000, size=n_eventos)

eventos = pd.DataFrame({
    "timestamp": timestamps,
    "tipo_evento": tipos_evento,
    "tamano_bytes": tamano_bytes,
})

print(eventos.head())
print("\nTotale eventi generati:", len(eventos))

# TODO 1 (VOLUME): calcola il totale di byte generati da tutti gli eventi insieme
# (somma la colonna 'tamano_bytes')
volumen_total_bytes = None

# TODO 2 (VELOCITÀ): calcola la finestra di tempo totale in minuti
# (sottrai il primo timestamp dall'ultimo, e converti il risultato in minuti con .total_seconds() / 60)
ventana_minutos = None

# TODO 3 (VELOCITÀ): calcola quanti eventi arrivano al minuto in media
# (numero totale di eventi diviso per ventana_minutos)
eventos_por_minuto = None

# TODO 4 (VARIETÀ): calcola quanti tipi diversi di evento ci sono e la loro distribuzione (value_counts)
variedad_tipos = None

print("\nVolume totale (byte):", volumen_total_bytes)
print("Finestra di tempo (minuti):", ventana_minutos)
print("Eventi al minuto (Velocità):", eventos_por_minuto)
print("\nVarietà di tipi di evento:")
print(variedad_tipos)
````
