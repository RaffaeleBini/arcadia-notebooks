```python
import pandas as pd

plan = pd.DataFrame({
    "habilidad": ["Python", "SQL", "Estadística", "Comunicación", "Curiosidad"],
    "nivel_actual": [2, 1, 3, 4, 5],
    "nivel_objetivo": [4, 4, 4, 4, 5],
})

plan["brecha"] = plan["nivel_objetivo"] - plan["nivel_actual"]

plan_ordenado = plan.sort_values("brecha", ascending=False)

print(plan_ordenado)
```

La competenza con la "brecha" maggiore è una buona prima candidata per il tuo piano di studio delle prossime settimane — non perché sia la più importante in assoluto, ma perché è dove una stessa ora di studio rende di più in termini di progresso relativo.
