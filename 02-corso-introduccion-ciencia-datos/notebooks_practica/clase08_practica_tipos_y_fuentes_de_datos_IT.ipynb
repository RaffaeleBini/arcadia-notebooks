import io

libros_json = '''
[
  {"titulo": "Cent anni di solitudine", "autor": "Garcia Marquez", "generos": ["realismo magico", "romanzo"]},

]
'''

libros_dict = json.loads(libros_json)
libros_df = pd.json_normalize(libros_dict)
print(libros_df)

libros_df["n_generos"] = libros_df["generos"].apply(len)
print(libros_df[["titulo", "n_generos"]])

```python
import pandas as pd
import io

empleados_csv = '''id_empleado,nombre,id_departamento,salario
1,Ana,10,32000
2,Luis,10,28000
3,Marta,20,41000
4,Jorge,20,39000
5,Sara,30,45000
'''

departamentos_json = '''
[
  {"id_departamento": 10, "nombre": "Marketing", "presupuesto": 90000},
  {"id_departamento": 20, "nombre": "Ingenieria", "presupuesto": 150000},
  {"id_departamento": 30, "nombre": "Ventas", "presupuesto": 60000}
]
'''

empleados = pd.read_csv(io.StringIO(empleados_csv))
departamentos = pd.read_json(io.StringIO(departamentos_json))

salarios_por_departamento = (
    empleados.groupby("id_departamento")["salario"].sum().reset_index()
)

resumen_departamentos = departamentos.merge(salarios_por_departamento, on="id_departamento")
resumen_departamentos["presupuesto_restante"] = (
    resumen_departamentos["presupuesto"] - resumen_departamentos["salario"]
)

print(resumen_departamentos[["nombre", "presupuesto", "salario", "presupuesto_restante"]])
```
