def clasificar_tarea(descripcion):
    """Classifica un compito come 'estadistica', 'data_scientist' o 'ambos'."""
    descripcion_lower = descripcion.lower()
    tiene_estadistica = any(palabra in descripcion_lower for palabra in palabras_estadistica)
    tiene_data_scientist = any(palabra in descripcion_lower for palabra in palabras_data_scientist)

    if tiene_estadistica and tiene_data_scientist:
        return "ambos"
    elif tiene_estadistica:
        return "estadistica"
    elif tiene_data_scientist:
        return "data_scientist"
    else:
        return "sin_clasificar"


for tarea in tareas:
    print(f"{tarea!r} -> {clasificar_tarea(tarea)}")
