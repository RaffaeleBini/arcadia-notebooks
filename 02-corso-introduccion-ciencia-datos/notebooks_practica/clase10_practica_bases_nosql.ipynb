## Ejercicio 2: ampliar la red del grafo

Sobre la simulación de grafo (`red_amistades`) de la clase de teoría, añade una persona nueva con al menos dos conexiones y vuelve a ejecutar `amigos_de_amigos` para comprobar cómo cambia el resultado.

**Pista:** solo necesitas añadir una nueva clave al diccionario `red_amistades` y, si quieres, añadir esa persona nueva como vecina de alguien que ya existe.

````python
red_amistades = {
    "Ana":   ["Luis", "Carla"],
    "Luis":  ["Ana", "Diego"],
    "Carla": ["Ana", "Marta"],
    "Diego": ["Luis"],
    "Marta": ["Carla"],
}

def amigos_de_amigos(grafo, persona):
    visitados = {persona}
    cola = [(vecino, 1) for vecino in grafo.get(persona, [])]
    resultado_nivel_2 = []
    while cola:
        actual, nivel = cola.pop(0)
        if actual in visitados:
            continue
        visitados.add(actual)
        if nivel == 2:
            resultado_nivel_2.append(actual)
            continue
        for vecino in grafo.get(actual, []):
            if vecino not in visitados:
                cola.append((vecino, nivel + 1))
    return resultado_nivel_2

# TODO: añade una persona nueva (por ejemplo "Sofia") a red_amistades
# con al menos dos conexiones, y añádela también como vecina de
# alguna persona que ya existe en el grafo.

# print(amigos_de_amigos(red_amistades, "Ana"))
````
