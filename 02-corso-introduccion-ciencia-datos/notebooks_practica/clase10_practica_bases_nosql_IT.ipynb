## Esercizio 2: ampliare la rete del grafo

Sulla simulazione di grafo (`red_amistades`) della lezione di teoria, aggiungi una nuova persona con almeno due connessioni e riesegui `amigos_de_amigos` per verificare come cambia il risultato.

**Suggerimento:** ti basta aggiungere una nuova chiave al dizionario `red_amistades` e, se vuoi, aggiungere quella nuova persona come vicina di qualcuno già esistente.

````python
red_amistades = {
    "Ana":   ["Luis", "Carla"],
    "Luis":  ["Ana", "Diego"],
    "Carla": ["Ana", "Marta"],
    "Diego": ["Luis"],
    "Marta": ["Carla"],
}

def amigos_de_amigos(grafo, persona):
    visitados = {persona}
    cola = [(vecino, 1) for vecino in grafo.get(persona, [])]
    resultado_nivel_2 = []
    while cola:
        actual, nivel = cola.pop(0)
        if actual in visitados:
            continue
        visitados.add(actual)
        if nivel == 2:
            resultado_nivel_2.append(actual)
            continue
        for vecino in grafo.get(actual, []):
            if vecino not in visitados:
                cola.append((vecino, nivel + 1))
    return resultado_nivel_2

# TODO: aggiungi una nuova persona (ad esempio "Sofia") a red_amistades
# con almeno due connessioni, e aggiungila anche come vicina di
# qualche persona già esistente nel grafo.

# print(amigos_de_amigos(red_amistades, "Ana"))
````
