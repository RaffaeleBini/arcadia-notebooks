## Ejercicio 2 — Prioriza tu plan de entrada

Vamos a construir una tabla con pandas que te ayude a decidir en qué habilidad enfocarte primero: la que tenga **mayor diferencia** entre tu nivel actual y tu nivel objetivo.

**Pistas:**
- `brecha = objetivo - actual` te dice cuánto te falta en cada habilidad.
- Ordena el DataFrame por esa columna, de mayor a menor, con `sort_values`.

````python
import pandas as pd

# Mismas habilidades que en el ejercicio 1, con un nivel actual y un nivel objetivo
# (el objetivo es el nivel que te gustaría alcanzar en, por ejemplo, 6 meses).
plan = pd.DataFrame({
    "habilidad": ["Python", "SQL", "Estadística", "Comunicación", "Curiosidad"],
    "nivel_actual": [2, 1, 3, 4, 5],
    "nivel_objetivo": [4, 4, 4, 4, 5],
})

# TODO: crea una columna "brecha" = nivel_objetivo - nivel_actual
plan["brecha"] = 0

# TODO: ordena el DataFrame "plan" por la columna "brecha", de mayor a menor,
# y guárdalo en una variable llamada "plan_ordenado"
plan_ordenado = plan

print(plan_ordenado)
````

<details>
<summary>💡 Ver solución</summary>

```python
import pandas as pd

plan = pd.DataFrame({
    "habilidad": ["Python", "SQL", "Estadística", "Comunicación", "Curiosidad"],
    "nivel_actual": [2, 1, 3, 4, 5],
    "nivel_objetivo": [4, 4, 4, 4, 5],
})

plan["brecha"] = plan["nivel_objetivo"] - plan["nivel_actual"]

plan_ordenado = plan.sort_values("brecha", ascending=False)

print(plan_ordenado)
```

La habilidad con mayor "brecha" es una buena primera candidata para tu plan de estudio de las próximas semanas — no porque sea la más importante en absoluto, sino porque es donde una misma hora de estudio rinde más en términos de progreso relativo.
