````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import numpy as np

diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target

print("Variables disponibles:", list(diabetes.feature_names))

# --- Modelo 1: bmi (igual que en clase, ya resuelto) ---
X_bmi = X[["bmi"]]
modelo_bmi = LinearRegression()
modelo_bmi.fit(X_bmi, y)
pred_bmi = modelo_bmi.predict(X_bmi)
error_bmi = mean_squared_error(y, pred_bmi)
print(f"MSE con 'bmi': {error_bmi:.2f}")

# --- Modelo 2: tu propia elección ---
# TODO: elige una variable de diabetes.feature_names distinta a "bmi"
otra_variable = "bmi"  # <- cámbiala por otra, por ejemplo "bp" o "s5"

X_otra = X[[otra_variable]]

# TODO: crea un LinearRegression, entrénalo con X_otra/y, y guarda
# las predicciones en `pred_otra`
modelo_otra = None
pred_otra = None

# TODO: calcula el error cuadrático medio de `pred_otra` frente a `y`
error_otra = None

print(f"MSE con '{otra_variable}': {error_otra}")
````
