## 9 · Manejo de errores

El bloque `try`/`except` captura una excepción sin detener el programa. Es imprescindible cuando conviertes texto introducido por quien juega en un número: si escribe algo que no es un número, `int(...)` lanza `ValueError`.


````python
def convertir_a_entero(texto):
    try:
        return int(texto)
    except ValueError:
        print(f"'{texto}' no es un número válido")
        return None

print(convertir_a_entero("3"))
print(convertir_a_entero("tres"))
````

## 10 · Ganador en horizontal

Hay tres formas de ganar en el Tic Tac Toe: en horizontal, en vertical y en diagonal. Empezamos por la horizontal: comprobar si alguna fila tiene sus casillas iguales y ocupadas.

El método `lista.count(valor)` cuenta cuántas veces aparece `valor` en la lista. Si el conteo de `"X"` en una fila coincide con la longitud de la fila, esa fila es toda de `"X"`.


````python
def ganador_horizontal(tablero):
    for fila in tablero:
        if fila.count("X") == len(fila):
            return "X"
        if fila.count("O") == len(fila):
            return "O"
    return None

tablero_x_gana = [["X", "X", "X"], ["O", "O", ""], ["", "", ""]]
print(ganador_horizontal(tablero_x_gana))   # X
print(ganador_horizontal(inicializar_tablero(3, 3)))   # None
````

## 13 · Iteradores e iterables

Un **iterable** es cualquier objeto que se puede recorrer con un bucle (listas, cadenas, tuplas...). Un **iterador** es el objeto que va entregando un elemento cada vez que lo llamas con `next()`, hasta agotarse y lanzar `StopIteration`.

Cuando escribes `for elemento in lista:`, Python está usando `iter()` y `next()` por debajo, de forma automática. Funciones como `zip()`, `enumerate()` o `reversed()` devuelven iteradores: si necesitas indexarlos o imprimirlos como una secuencia, conviértelos antes con `list(...)`.


````python
numeros = [10, 20, 30]
iterador = iter(numeros)
print(next(iterador))   # 10
print(next(iterador))   # 20
print(next(iterador))   # 30

# zip combina dos iterables en pares
filas = range(3)
columnas_invertidas = reversed(range(3))
for fila, columna in zip(filas, columnas_invertidas):
    print(fila, columna)
````

## 16 · El bucle principal: el juego completo

Ya tienes todas las piezas: crear el tablero, mostrarlo, validar movimientos, pedir números por teclado con control de errores, y detectar ganador o empate. Solo falta el bucle que gestiona los turnos.

La función `jugar()` de abajo:

1. Crea un tablero 3×3
2. En cada turno, muestra el tablero y pide fila y columna a quien le toca jugar (usando `pedir_entero`)
3. Si el movimiento no es válido, vuelve a pedirlo sin cambiar de turno
4. Si el movimiento es válido, coloca el símbolo, comprueba si hay ganador o empate, y si no, cambia de turno
5. Al terminar la partida, pregunta si quieren jugar de nuevo


````python
def jugar():
    tamano = 3
    tablero = inicializar_tablero(tamano, tamano)
    turnos = ["X", "O"]
    turno_actual = 0

    while True:
        mostrar_tablero(tablero)
        simbolo = turnos[turno_actual]
        print(f"\nTurno de: {simbolo}")

        fila = pedir_entero(f"Fila (0-{tamano - 1}): ")
        columna = pedir_entero(f"Columna (0-{tamano - 1}): ")

        if not movimiento_valido(tablero, fila, columna):
            print("Movimiento no válido, prueba con otra casilla.\n")
            continue

        tablero[fila][columna] = simbolo

        resultado = ganador(tablero)
        if resultado:
            simbolo_ganador, tipo = resultado
            mostrar_tablero(tablero)
            print(f"\n¡{simbolo_ganador} ha ganado en {tipo}!")
            break

        if tablero_lleno(tablero):
            mostrar_tablero(tablero)
            print("\n¡Empate! El tablero se ha llenado sin ganador.")
            break

        turno_actual = 1 - turno_actual

    respuesta = input("\n¿Jugar otra partida? (s/n): ").strip().lower()
    if respuesta == "s":
        jugar()
    elif respuesta == "n":
        print("¡Gracias por jugar!")
    else:
        print("Respuesta no válida. Hasta la próxima.")
````
