## 9 · Gestione degli errori

Il blocco `try`/`except` cattura un'eccezione senza fermare il programma. È essenziale quando converti il testo inserito da chi gioca in un numero: se scrive qualcosa che non è un numero, `int(...)` lancia `ValueError`.


````python
def convertir_a_entero(texto):
    try:
        return int(texto)
    except ValueError:
        print(f"'{texto}' no es un número válido")
        return None

print(convertir_a_entero("3"))
print(convertir_a_entero("tres"))
````

## 10 · Vincitore in orizzontale

Hay tres formas de ganar nel Tic Tac Toe: in orizzontale, verticale e diagonale. Iniziamo con l'orizzontale: controllare se qualche riga ha le sue caselle uguali e occupate.

Il metodo `lista.count(valore)` conta quante volte appare `valore` nella lista. Se il conteggio di `"X"` in una riga coincide con la lunghezza della riga, quella riga è tutta di `"X"`.


````python
def ganador_horizontal(tablero):
    for fila in tablero:
        if fila.count("X") == len(fila):
            return "X"
        if fila.count("O") == len(fila):
            return "O"
    return None

tablero_x_gana = [["X", "X", "X"], ["O", "O", ""], ["", "", ""]]
print(ganador_horizontal(tablero_x_gana))   # X
print(ganador_horizontal(inicializar_tablero(3, 3)))   # None
````

## 13 · Iteratori e iterabili

Un **iterable** es cualquier objeto che si può percorrere con un loop (listas, cadenas, tuplas...). Un **iterador** es el objeto que va entregando un elemento ogni volta che lo chiami con `next()`, hasta agotarse y lanzar `StopIteration`.

Quando scrivi `for elemento in lista:`, Python sta usando `iter()` e `next()` dietro le quinte, automaticamente. Funzioni come `zip()`, `enumerate()` o `reversed()` restituiscono iteratori: se hai bisogno di indicizzarli o stamparli come una sequenza, convertili prima con `list(...)`.


````python
numeros = [10, 20, 30]
iterador = iter(numeros)
print(next(iterador))   # 10
print(next(iterador))   # 20
print(next(iterador))   # 30

# zip combina dos iterables en pares
filas = range(3)
columnas_invertidas = reversed(range(3))
for fila, columna in zip(filas, columnas_invertidas):
    print(fila, columna)
````

## 16 · Il loop principale: il gioco completo

Hai già tutti i pezzi: creare la tavola, mostrarla, validare i movimenti, chiedere numeri da tastiera con controllo degli errori, e rilevare vincitore o pareggio. Manca solo il loop che gestisce i turni.

La funzione `giocare()` qui sotto:

1. Crea una tavola 3×3
2. In ogni turno, mostra la tavola e chiedi riga e colonna a chi tocca giocare (usando `chiedere_intero`)
3. Se il movimento non è valido, lo chiedi di nuovo senza cambiare turno
4. Se il movimento è valido, piazza il simbolo, controlla se c'è un vincitore o un pareggio, e se no, cambia turno
5. Alla fine della partita, chiedi se vogliono giocare di nuovo


````python
def jugar():
    tamano = 3
    tablero = inicializar_tablero(tamano, tamano)
    turnos = ["X", "O"]
    turno_actual = 0

    while True:
        mostrar_tablero(tablero)
        simbolo = turnos[turno_actual]
        print(f"\nTurno de: {simbolo}")

        fila = pedir_entero(f"Fila (0-{tamano - 1}): ")
        columna = pedir_entero(f"Columna (0-{tamano - 1}): ")

        if not movimiento_valido(tablero, fila, columna):
            print("Movimiento no válido, prueba con otra casilla.\n")
            continue

        tablero[fila][columna] = simbolo

        resultado = ganador(tablero)
        if resultado:
            simbolo_ganador, tipo = resultado
            mostrar_tablero(tablero)
            print(f"\n¡{simbolo_ganador} ha ganado en {tipo}!")
            break

        if tablero_lleno(tablero):
            mostrar_tablero(tablero)
            print("\n¡Empate! El tablero se ha llenado sin ganador.")
            break

        turno_actual = 1 - turno_actual

    respuesta = input("\n¿Jugar otra partida? (s/n): ").strip().lower()
    if respuesta == "s":
        jugar()
    elif respuesta == "n":
        print("¡Gracias por jugar!")
    else:
        print("Respuesta no válida. Hasta la próxima.")
````
