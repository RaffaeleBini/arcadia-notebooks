## 1. Variables: cajas con nombre

Una **variable** es una caja con un nombre donde guardas un valor para usarlo después. Se crea con `=` (que aquí significa "guarda esto", no "es igual a").

Los tipos de dato básicos que usaremos:
- **Texto** (*string*): va entre comillas, p. ej. `"Madrid"`
- **Número entero** (*int*): p. ej. `42`
- **Número decimal** (*float*): p. ej. `19.99`
- **Booleano** (*bool*): solo `True` o `False`

```python
nombre = "Ana"          # texto
edad = 34               # numero entero
altura = 1.68           # numero decimal
es_cliente = True       # booleano

# print() muestra valores por pantalla. Las comas separan varios valores.
print(nombre, "tiene", edad, "años y mide", altura, "m")
print("¿Es cliente?", es_cliente)

# type() te dice de qué tipo es cada valor
print("El tipo de 'edad' es:", type(edad))
```

### Operaciones básicas

Con los números puedes operar como en una calculadora. Con el texto, puedes **unirlo** con `+`.

```python
precio = 100
descuento = 0.20                       # 20%
precio_final = precio * (1 - descuento)
print("Precio final:", precio_final, "€")

saludo = "Hola, " + nombre + "!"        # unir textos se llama 'concatenar'
print(saludo)
```

## 2. Listas: varios valores en una sola variable

Una **lista** guarda varios elementos en orden, entre corchetes `[]`. Es la estructura más usada para manejar conjuntos de datos.

Detalle importante: en Python **se cuenta desde 0**. El primer elemento es la posición `0`, el segundo la `1`, etc.

```python
ventas = [120, 340, 90, 500, 210]

print("Lista completa:", ventas)
print("Primer elemento (posición 0):", ventas[0])
print("Último elemento:", ventas[-1])     # -1 = el último
print("Cuántos elementos hay:", len(ventas))

# Añadir un elemento al final
ventas.append(75)
print("Tras añadir 75:", ventas)

# Funciones útiles sobre listas de números
print("Suma:", sum(ventas), "| Máximo:", max(ventas), "| Mínimo:", min(ventas))
```

## 3. Condicionales: tomar decisiones (`if`)

Un condicional ejecuta un bloque de código **solo si** se cumple una condición. La **sangría** (los espacios al principio de la línea) es la que indica qué está "dentro" del `if`. En Python la sangría no es decorativa: es obligatoria.

```python
temperatura = 14.5

if temperatura > 25:
    print("Hace calor")
elif temperatura > 12:                 # 'elif' = "si no, y si..."
    print("Temperatura agradable")
else:
    print("Hace frío")
```

## 4. Bucles: repetir sin copiar y pegar

Un **bucle `for`** repite un bloque de código para cada elemento de una lista. Es una de las herramientas que más usarás.

```python
ventas = [120, 340, 90, 500, 210]

# Recorrer la lista elemento a elemento
for venta in ventas:
    if venta >= 300:
        print(venta, "→ venta grande ✅")
    else:
        print(venta, "→ venta normal")
```

## 5. Funciones: empaquetar código para reutilizarlo

Una **función** es un bloque de código con nombre que puedes reutilizar cuantas veces quieras. Se define con `def`, recibe **parámetros** (entradas) y devuelve un resultado con `return`.

Ya has usado funciones: `print()`, `len()` y `sum()` son funciones. Ahora crearás las tuyas.

```python
def precio_con_iva(precio, iva=0.21):
    """Devuelve el precio sumándole el IVA (21% por defecto)."""
    return precio * (1 + iva)

# Llamar (usar) la función
print(precio_con_iva(100))        # usa el 21% por defecto -> 121.0
print(precio_con_iva(100, 0.10))  # IVA del 10% -> 110.0
```

## 6. Librerías clave: numpy y pandas

No tienes que reinventar la rueda. Una **librería** es código ya hecho por otras personas que importas y usas. Para datos e IA, dos son imprescindibles:

- **numpy** — cálculo numérico rápido con vectores y matrices. La base sobre la que se construye casi todo
- **pandas** — manejo de datos en forma de **tabla** (filas y columnas), como una hoja de cálculo pero en código

> En Google Colab ya vienen instaladas. Si trabajas en tu ordenador y faltan, instálalas con `pip install numpy pandas` en la terminal.

```python
import numpy as np      # 'np' y 'pd' son apodos estándar que usa todo el mundo
import pandas as pd

# numpy: operar sobre muchos números a la vez, sin bucles
temperaturas = np.array([14.5, 18.0, 22.3, 9.8, 30.1])
print("Media:", temperaturas.mean().round(2))
print("Máxima:", temperaturas.max())
print("Todas +2 grados:", temperaturas + 2)   # suma a TODOS de golpe
```

### pandas: el DataFrame, tu tabla de datos

La estructura estrella de pandas es el **DataFrame**: una tabla con filas y columnas. Vamos a crear una a mano para verla.

```python
datos = {
    "ciudad": ["Santiago", "Vigo", "Coruña", "Lugo"],
    "habitantes": [97000, 293000, 245000, 98000],
    "lluvia_anual_mm": [1900, 1900, 1000, 900],
}
df = pd.DataFrame(datos)
df    # al poner el nombre del DataFrame solo, el notebook lo muestra como tabla
```

```python
# Operaciones habituales sobre un DataFrame
print("Columna de habitantes:")
print(df["habitantes"])
print("\nMedia de lluvia:", df["lluvia_anual_mm"].mean())
print("\nFiltrar ciudades con más de 100.000 habitantes:")
df[df["habitantes"] > 100000]
```

## 7. Leer y explorar un dataset real

Hasta ahora inventamos los datos. Vamos a cargar un **conjunto de datos real y muy conocido**: las medidas de tres especies de la flor *iris*. Viene incluido en la librería `scikit-learn`, así que **no necesita internet**.

> En la vida real cargarás tus propios datos desde un archivo con `pd.read_csv("mis_datos.csv")`. La forma de explorarlos es exactamente la misma que verás aquí.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame.copy()
# Añadimos el nombre de la especie (más legible que un número)
df["especie"] = df["target"].map(dict(enumerate(iris.target_names)))

print("Forma de la tabla (filas, columnas):", df.shape)
df.head()     # .head() muestra las 5 primeras filas
```

Cuatro comandos que usarás **siempre** al recibir datos nuevos. Memorízalos:

```python
# .info() te dice cuántas filas hay, las columnas y si faltan datos
df.info()
```

```python
# .describe() da un resumen estadístico de las columnas numéricas
df.describe().round(2)
```

```python
# .value_counts() cuenta cuántas veces aparece cada valor de una columna
df["especie"].value_counts()
```

```python
# .groupby() agrupa y resume: aquí, el largo medio de pétalo por especie
df.groupby("especie")["petal length (cm)"].mean().round(2)
```
