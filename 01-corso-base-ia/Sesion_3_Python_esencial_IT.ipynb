## 1. Variabili: scatole con un nome

Una **variabile** è una scatola con un nome dove salvi un valore per usarlo dopo. Si crea con `=` (che qui significa "salva questo", non "è uguale a").

I tipi di dato di base che useremo:
- **Testo** (*string*): va tra virgolette, es. `"Madrid"`
- **Numero intero** (*int*): es. `42`
- **Numero decimale** (*float*): es. `19.99`
- **Booleano** (*bool*): solo `True` o `False`

```python
nombre = "Ana"          # testo
edad = 34               # numero intero
altura = 1.68           # numero decimale
es_cliente = True       # booleano

# print() mostra i valori a schermo. Le virgole separano più valori.
print(nombre, "ha", edad, "anni ed è alta", altura, "m")
print("È cliente?", es_cliente)

# type() ti dice di che tipo è ogni valore
print("Il tipo di 'edad' è:", type(edad))
```

### Operazioni di base

Con i numeri puoi operare come con una calcolatrice. Con il testo, puoi **unirlo** con `+`.

```python
precio = 100
descuento = 0.20                       # 20%
precio_final = precio * (1 - descuento)
print("Prezzo finale:", precio_final, "€")

saludo = "Ciao, " + nombre + "!"        # unire testi si chiama 'concatenare'
print(saludo)
```

## 2. Liste: più valori in una sola variabile

Una **lista** salva più elementi in ordine, tra parentesi quadre `[]`. È la struttura più usata per gestire insiemi di dati.

Dettaglio importante: in Python **si conta da 0**. Il primo elemento è la posizione `0`, il secondo la `1`, ecc.

```python
ventas = [120, 340, 90, 500, 210]

print("Lista completa:", ventas)
print("Primo elemento (posizione 0):", ventas[0])
print("Ultimo elemento:", ventas[-1])     # -1 = l'ultimo
print("Quanti elementi ci sono:", len(ventas))

# Aggiungere un elemento in fondo
ventas.append(75)
print("Dopo aver aggiunto 75:", ventas)

# Funzioni utili sulle liste di numeri
print("Somma:", sum(ventas), "| Massimo:", max(ventas), "| Minimo:", min(ventas))
```

## 3. Condizionali: prendere decisioni (`if`)

Un condizionale esegue un blocco di codice **solo se** una condizione è vera. L'**indentazione** (gli spazi all'inizio della riga) è ciò che indica cosa sta "dentro" l'`if`. In Python l'indentazione non è decorativa: è obbligatoria.

```python
temperatura = 14.5

if temperatura > 25:
    print("Fa caldo")
elif temperatura > 12:                 # 'elif' = "altrimenti, se..."
    print("Temperatura piacevole")
else:
    print("Fa freddo")
```

## 4. Cicli: ripetere senza copiare e incollare

Un **ciclo `for`** ripete un blocco di codice per ogni elemento di una lista. È uno degli strumenti che userai di più.

```python
ventas = [120, 340, 90, 500, 210]

# Scorrere la lista elemento per elemento
for venta in ventas:
    if venta >= 300:
        print(venta, "→ vendita grande ✅")
    else:
        print(venta, "→ vendita normale")
```

## 5. Funzioni: impacchettare codice per riusarlo

Una **funzione** è un blocco di codice con un nome che puoi riutilizzare tutte le volte che vuoi. Si definisce con `def`, riceve dei **parametri** (input) e restituisce un risultato con `return`.

Hai già usato delle funzioni: `print()`, `len()` e `sum()` sono funzioni. Ora creerai le tue.

```python
def precio_con_iva(precio, iva=0.21):
    """Restituisce il prezzo con l'IVA aggiunta (21% di default)."""
    return precio * (1 + iva)

# Chiamare (usare) la funzione
print(precio_con_iva(100))        # usa il 21% di default -> 121.0
print(precio_con_iva(100, 0.10))  # IVA al 10% -> 110.0
```

## 6. Librerie chiave: numpy e pandas

Non devi reinventare la ruota. Una **libreria** è codice già scritto da altre persone che importi e usi. Per dati e IA, due sono indispensabili:

- **numpy** — calcolo numerico veloce con vettori e matrici. La base su cui si costruisce quasi tutto
- **pandas** — gestione dei dati in forma di **tabella** (righe e colonne), come un foglio di calcolo ma in codice

> Su Google Colab sono già installate. Se lavori sul tuo computer e mancano, installale con `pip install numpy pandas` nel terminale.

```python
import numpy as np      # 'np' e 'pd' sono i soprannomi standard usati da tutti
import pandas as pd

# numpy: operare su molti numeri insieme, senza cicli
temperaturas = np.array([14.5, 18.0, 22.3, 9.8, 30.1])
print("Media:", temperaturas.mean().round(2))
print("Massima:", temperaturas.max())
print("Tutte +2 gradi:", temperaturas + 2)   # somma a TUTTI in un colpo solo
```

### pandas: il DataFrame, la tua tabella di dati

La struttura principe di pandas è il **DataFrame**: una tabella con righe e colonne. Ne creiamo una a mano per vederla.

```python
datos = {
    "ciudad": ["Santiago", "Vigo", "Coruña", "Lugo"],
    "habitantes": [97000, 293000, 245000, 98000],
    "lluvia_anual_mm": [1900, 1900, 1000, 900],
}
df = pd.DataFrame(datos)
df    # scrivendo solo il nome del DataFrame, il notebook lo mostra come tabella
```

```python
# Operazioni comuni su un DataFrame
print("Colonna degli abitanti:")
print(df["habitantes"])
print("\nMedia della pioggia:", df["lluvia_anual_mm"].mean())
print("\nFiltrare le città con più di 100.000 abitanti:")
df[df["habitantes"] > 100000]
```

## 7. Leggere ed esplorare un dataset reale

Finora abbiamo inventato i dati. Carichiamo un **dataset reale e molto conosciuto**: le misure di tre specie del fiore *iris*. È incluso nella libreria `scikit-learn`, quindi **non serve internet**.

> Nella vita reale caricherai i tuoi dati da un file con `pd.read_csv("mis_datos.csv")`. Il modo di esplorarli è esattamente lo stesso che vedrai qui.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame.copy()
# Aggiungiamo il nome della specie (più leggibile di un numero)
df["especie"] = df["target"].map(dict(enumerate(iris.target_names)))

print("Forma della tabella (righe, colonne):", df.shape)
df.head()     # .head() mostra le prime 5 righe
```

Quattro comandi che userai **sempre** quando ricevi nuovi dati. Memorizzali:

```python
# .info() ti dice quante righe ci sono, le colonne e se mancano dati
df.info()
```

```python
# .describe() dà un riepilogo statistico delle colonne numeriche
df.describe().round(2)
```

```python
# .value_counts() conta quante volte compare ogni valore di una colonna
df["especie"].value_counts()
```

```python
# .groupby() raggruppa e riassume: qui, la lunghezza media del petalo per specie
df.groupby("especie")["petal length (cm)"].mean().round(2)
```
