## 1. Dati, modello e predizione

Tre parole che sentirai mille volte. Fissiamole con un'immagine semplice:

- **Dati** — gli esempi da cui impara la macchina. Ad esempio, schede di abitazioni con i loro metri quadri, il quartiere e il prezzo finale
- **Modello** — ciò che la macchina "costruisce" dopo aver osservato i dati. È come una **ricetta imparata**: a partire da alcuni ingredienti (i dati in ingresso) produce un risultato (la predizione)
- **Predizione** — la risposta del modello di fronte a un caso nuovo che non aveva visto

Il flusso è sempre lo stesso:

> **Dati di esempio → (addestramento) → Modello → (predizione) → Risposta su un caso nuovo**

Un dettaglio di vocabolario molto utile: nei dati chiamiamo **caratteristiche** (o *feature*) le informazioni in ingresso (metri quadri, quartiere...) ed **etichetta** (o *target*) ciò che vogliamo predire (il prezzo).

## 2. Addestrare e validare: studiare e sostenere l'esame

Ecco una delle idee più importanti e più facili da dimenticare.

Quando una macchina impara, **non possiamo giudicarla con gli stessi dati che ha usato per imparare**. Sarebbe come darti l'esame con le stesse domande (e risposte) che avevi già per studiare: prenderesti il massimo dei voti senza aver dimostrato di aver imparato nulla.

Per questo dividiamo i dati in due parti:

- **Insieme di addestramento** (*training*) — gli esempi con cui il modello impara. È il materiale di studio
- **Insieme di validazione o test** (*test*) — esempi che il modello **non ha visto** e che usiamo per valutarlo. È l'esame vero e proprio

Una divisione tipica è lasciare l'80% dei dati per l'addestramento e il 20% per la valutazione. Ciò che conta davvero non è quanto bene il modello risponde su ciò che ha già studiato, ma **come si comporta di fronte a dati nuovi**. Questo lo chiamiamo *generalizzare*.

## 3. Sovradattamento e sottoadattamento: l'equilibrio dell'apprendimento

Quando un modello impara, può sbagliare in due modi opposti. Te li spiego con l'esempio di chi studia per un esame.

**Sottoadattamento (underfitting) — ha studiato troppo poco**
Il modello è troppo semplice e non coglie nemmeno i pattern evidenti. Fallisce sia nello studio sia all'esame.
- Segnale tipico: funziona male sia con i dati di addestramento *sia* con quelli di test

**Sovradattamento (overfitting) — ha memorizzato invece di capire**
Il modello impara a memoria gli esempi, incluso il "rumore" e le coincidenze. Eccelle sul materiale di studio, ma crolla di fronte a una domanda nuova.
- Segnale tipico: funziona quasi alla perfezione in addestramento, ma male nel test

**Il punto giusto** sta nel mezzo: un modello che coglie i pattern reali senza memorizzare le coincidenze. Buona parte del lavoro nel ML consiste nel trovare questo equilibrio.

### 🖥️ Demo opzionale del docente (senza codice per gli studenti)

Questa cella disegna i tre casi sugli stessi punti. **Non devi capirla**, limitati a eseguirla per vedere la differenza mentre la spiego.

```python
# Demo visiva: sottoadattamento vs. adattamento corretto vs. sovradattamento
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
x = np.linspace(0, 1, 18)
y = np.sin(2 * np.pi * x) + rng.normal(0, 0.25, x.size)   # pattern reale + rumore
xx = np.linspace(0, 1, 200)

fig, ax = plt.subplots(1, 3, figsize=(13, 4))
for i, (grado, titulo) in enumerate([(1, "Sottoadattamento\n(troppo semplice)"),
                                      (4, "Adattamento corretto\n(coglie il pattern)"),
                                      (14, "Sovradattamento\n(memorizza il rumore)")]):
    coef = np.polyfit(x, y, grado)
    ax[i].scatter(x, y, color="#1f77b4", label="dati")
    ax[i].plot(xx, np.polyval(coef, xx), color="#d62728", lw=2, label="modello")
    ax[i].set_title(titulo); ax[i].set_ylim(-2, 2); ax[i].legend(loc="upper right", fontsize=8)
plt.tight_layout(); plt.show()
```

## 4. Reti neurali senza matematica

Le reti neurali sono il cuore del Deep Learning e dei modelli di linguaggio. Cerchiamo di capirle **senza una sola formula**.

**Il neurone artificiale.** Immagina una piccola calcolatrice che riceve diversi numeri in ingresso, dà più o meno importanza a ciascuno (questi sono i **pesi**) e produce un numero in uscita. Un singolo neurone è molto limitato.

**La rete.** Il bello sta nel connettere **molti** neuroni in **strati**: l'informazione entra da uno strato, passa per strati intermedi (i "nascosti") ed esce dall'ultimo. Ogni strato trasforma l'informazione in qualcosa di più utile.

**E l'apprendimento?** All'inizio i pesi sono casuali, quindi la rete sbaglia. Ogni volta che sbaglia, **regola un po' i suoi pesi** per sbagliare meno la volta successiva. Ripete questo processo milioni di volte. È come affinare a poco a poco migliaia di manopole di un mixer audio finché il suono non è quello giusto.

**L'idea essenziale:** una rete neurale non memorizza regole; memorizza **un gran numero di pesi regolati** che, insieme, catturano i pattern dei dati. Quando senti dire che un modello "ha 7 miliardi di parametri", quei parametri sono proprio quei pesi.

### Perché "profonda"?

Quando una rete ha **molti strati**, diciamo che è *profonda*: da qui viene *Deep Learning* (apprendimento profondo).

Più strati permettono di imparare pattern per livelli. In un'immagine, ad esempio, i primi strati individuano bordi semplici, quelli successivi forme, e gli ultimi oggetti completi come "un occhio" o "una ruota". Nessuno ha programmato quella gerarchia: **emerge da sola dall'addestramento**.
