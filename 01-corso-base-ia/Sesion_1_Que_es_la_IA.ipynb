## 1. ¿Qué es la Inteligencia Artificial?

Empecemos sin tecnicismos. **La Inteligencia Artificial (IA) es la disciplina que busca que las máquinas hagan tareas que, cuando las hace una persona, asociamos con la inteligencia**: reconocer una cara, entender una frase, conducir, recomendar una película o detectar un fraude.

Fíjate en un matiz importante: la IA no "piensa" como tú. No tiene intenciones ni comprende el mundo. Lo que hace es **encontrar patrones** en muchos ejemplos y usarlos para responder ante situaciones nuevas.

Una analogía útil: imagina a alguien que nunca ha visto una manzana. Le enseñas mil fotos diciéndole "esto es una manzana / esto no lo es". Al final, sin que le expliques ninguna regla, aprende a reconocerlas. La IA funciona parecido: **aprende de ejemplos, no de reglas que le dictamos una a una.**

### La distinción clave: programación tradicional vs. IA

Esta es probablemente la idea más importante de toda la sesión.

| | Programación tradicional | Inteligencia Artificial |
|---|---|---|
| Tú aportas | Las **reglas** + los datos | Los **datos** + las respuestas correctas |
| La máquina produce | Las respuestas | Las **reglas** (el modelo) |
| Ejemplo | "Si el importe > 1.000 €, marca como sospechoso" | Le muestras miles de fraudes y aprende sola qué los caracteriza |

En la programación de siempre, una persona escribe las reglas. En la IA, **la máquina deduce las reglas a partir de los ejemplos**. Por eso la IA brilla en problemas donde las reglas son demasiado complejas para escribirlas a mano: ¿cómo describirías con reglas exactas qué hace que una cara sea "la de tu madre"?

## 2. IA, ML y Deep Learning: no son lo mismo

Estos tres términos se usan como sinónimos en los titulares, pero **uno está dentro del otro**, como muñecas rusas:

- **Inteligencia Artificial (IA)** — el campo más amplio. Toda técnica que da a las máquinas comportamientos "inteligentes". Incluye cosas tan antiguas como los sistemas de reglas de los años 80.
- **Aprendizaje Automático (Machine Learning, ML)** — un subconjunto de la IA. Aquí la máquina **aprende de los datos** en lugar de seguir reglas escritas a mano. Es el motor de casi toda la IA útil de hoy.
- **Aprendizaje Profundo (Deep Learning, DL)** — un subconjunto del ML que usa **redes neuronales con muchas capas**. Es lo que hay detrás del reconocimiento de voz, la visión por ordenador y los modelos de lenguaje como Claude o GPT.

En resumen: **todo Deep Learning es ML, y todo ML es IA, pero no al revés.**

## 3. Una historia muy breve (para situarte)

No hace falta memorizar fechas, pero conviene saber que la IA **no es nueva**: lleva décadas con nosotros, con épocas de euforia y de frustración.

- **1950** — Alan Turing se pregunta "¿pueden pensar las máquinas?" y propone el famoso *test de Turing*
- **1956** — En la conferencia de Dartmouth se acuña el término "Inteligencia Artificial"
- **Años 80** — Auge de los *sistemas expertos*: IA basada en reglas escritas por personas
- **2012** — Una red neuronal profunda (AlexNet) gana por goleada un concurso de visión y reabre la era del Deep Learning
- **2017** — Aparece la arquitectura *Transformer* (artículo "Attention is all you need"), base de los modelos de lenguaje actuales
- **2022 en adelante** — Los modelos de lenguaje conversacionales llegan al gran público y la IA se vuelve cotidiana

La lección de fondo: **lo que cambió no fue solo la idea, sino tres cosas que se juntaron**: muchos más datos, ordenadores mucho más potentes y mejores algoritmos.

## 4. Los tres grandes tipos de aprendizaje

Casi todo el ML encaja en una de estas tres familias. Te las explico con analogías cotidianas.

**Aprendizaje supervisado** — aprendes con un profesor que te corrige. Le das a la máquina ejemplos *con la respuesta correcta* ("esta foto es un gato", "este correo es spam"). Es el más usado en la práctica.
- Ejemplos: detectar spam, predecir el precio de una vivienda, diagnosticar a partir de una radiografía

**Aprendizaje no supervisado** — aprendes sin profesor, buscando estructura por tu cuenta. Le das datos *sin respuestas* y la máquina agrupa lo que se parece.
- Ejemplos: segmentar clientes en grupos, detectar anomalías, organizar noticias por temas

**Aprendizaje por refuerzo** — aprendes por ensayo y error, como quien aprende a montar en bici. La máquina prueba acciones y recibe premios o castigos.
- Ejemplos: un robot que aprende a caminar, una IA que juega al ajedrez, optimización de rutas

## 5. Casos reales de uso (probablemente ya los usas)

La IA ya está en tu día a día, casi siempre sin que la veas:

- **Salud y cuidados** — apoyo al diagnóstico por imagen, priorización de pacientes, detección temprana
- **Agroalimentación y mar** — control de calidad por visión, predicción de cosechas, trazabilidad
- **Forestal y madera** — detección de plagas, optimización del uso de recursos
- **Industria** — mantenimiento predictivo, control de defectos en la línea de producción
- **Vida cotidiana** — recomendaciones, traducción automática, asistentes de voz, antifraude bancario

Idea para llevarte a casa: la IA no es magia ni ciencia ficción. **Es una herramienta para encontrar patrones en datos**, y su valor aparece cuando la aplicas a un problema concreto y bien definido.
