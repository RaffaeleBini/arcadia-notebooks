# Introducción
Este curso está diseñado para personas que parten desde cero y quieren llegar a construir sus propias aplicaciones de inteligencia artificial.

No necesitas ningún conocimiento de programación previo: empezamos desde cero.

Para lograr este objetivo, procederemos paso a paso por un camino guiado estructurado en 4 módulos.

**Estructura en 4 módulos**. Está pensada para que haya una progresión natural: primero entender la IA sin código, luego aprender las herramientas técnicas, después usar modelos reales, y finalmente crear algo propio.

**Metodología**. Cada sesión combina teoría (20 min) + práctica guiada (30 min) + recap y preguntas (10 min). Entre clases se deja un ejercicio corto para afianzar lo visto.

El programa completo son 8 sesiones de 60 minutos, organizadas en pares: cada módulo ocupa exactamente 2 clases, lo que permite introducir el tema en la primera y practicarlo en la segunda.
Un detalle importante sobre la progresión: los módulos 1 y 2 son la base técnica y conceptual; los módulos 3 y 4 son 100% aplicados. Así llegarás al proyecto final (módulo 4) con todo lo que necesitas para construir algo real.
