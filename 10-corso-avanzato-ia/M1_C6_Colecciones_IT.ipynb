## 1. Liste (`list`)

Una lista è una sequenza **ordinata** e **mutabile** di elementi. È la collezione più usata in Python.

- **Ordinata:** gli elementi hanno una posizione fissa (indice).
- **Mutabile:** puoi aggiungere, eliminare o modificare elementi dopo averla creata.
- **Ammette duplicati:** lo stesso valore può comparire più volte.
- **Eterogenea:** può contenere elementi di tipi diversi.

```python
voti        = [8.5, 6.0, 9.0, 5.5]         # lista di float
nomi        = ["Ana", "Luis", "Maria"]      # lista di string
mista       = [1, "ciao", True, 3.14]       # tipi misti
vuota       = []                             # lista vuota
annidata    = [[1, 2], [3, 4], [5, 6]]      # lista di liste
```

````python
# Creare liste
voti = [8.5, 6.0, 9.0, 5.5, 7.0]
nomi = ["Ana", "Luis", "Maria", "Carlos"]
vuota = []

print(voti)
print(nomi)
print(type(voti))       # <class 'list'>
print(len(voti))        # numero di elementi
````

### 1.1 Accesso per indice e slicing

Come per le string, le liste usano indici a partire da `0`. Gli indici negativi contano a partire dalla fine.

```
lista  =  ["a",  "b",  "c",  "d",  "e"]
indice:     0     1     2     3     4
negativo:  -5    -4    -3    -2    -1
```

Lo **slicing** `lista[inizio:fine:passo]` restituisce una sottolista (l'indice `fine` non è incluso).

````python
frutta = ["mela", "pera", "uva", "arancia", "kiwi"]

# Accesso per indice
print(frutta[0])    # mela
print(frutta[-1])   # kiwi (ultimo)
print(frutta[2])    # uva

# Slicing
print(frutta[1:3])  # ['pera', 'uva']
print(frutta[:2])   # ['mela', 'pera']
print(frutta[3:])   # ['arancia', 'kiwi']
print(frutta[::-1]) # lista al contrario
````

### 1.2 Modificare le liste

Le liste sono **mutabili**: puoi modificare, aggiungere ed eliminare elementi.

| Operazione | Metodo / Sintassi | Cosa fa? |
|-----------|------------------|------------|
| Modificare un elemento | `lista[i] = valore` | Sostituisce l'elemento in posizione `i` |
| Aggiungere in fondo | `lista.append(x)` | Aggiunge `x` in fondo |
| Inserire in posizione | `lista.insert(i, x)` | Inserisce `x` in posizione `i` |
| Aggiungere un'altra lista | `lista.extend(altra)` | Aggiunge tutti gli elementi di `altra` |
| Eliminare per valore | `lista.remove(x)` | Elimina la prima occorrenza di `x` |
| Eliminare per indice | `lista.pop(i)` | Elimina e restituisce l'elemento in `i` |
| Svuotare | `lista.clear()` | Elimina tutti gli elementi |
| Eliminare (senza ritorno) | `del lista[i]` | Elimina l'elemento in `i` |

````python
colori = ["rosso", "verde", "blu"]
print(f"Iniziale:          {colori}")

colori.append("giallo")
print(f"Dopo append:       {colori}")

colori.insert(1, "arancione")
print(f"Dopo insert(1):    {colori}")

colori.remove("verde")
print(f"Dopo remove:       {colori}")

eliminato = colori.pop(0)
print(f"Dopo pop(0):       {colori}  ← eliminato: '{eliminato}'")

colori[0] = "viola"
print(f"Dopo modifica [0]: {colori}")
````

### 1.3 Altri metodi e operazioni utili

````python
numeri = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

print(f"Lista originale:  {numeri}")
print(f"len():            {len(numeri)}")
print(f"sum():            {sum(numeri)}")
print(f"max():            {max(numeri)}")
print(f"min():            {min(numeri)}")
print(f"count(5):         {numeri.count(5)}")   # quante volte compare il 5
print(f"index(9):         {numeri.index(9)}")   # posizione della prima occorrenza
print(f"'4' in lista:     {4 in numeri}")

# Ordinare
copia = numeri.copy()   # copia per non modificare l'originale
copia.sort()
print(f"sorted (asc):     {copia}")
copia.sort(reverse=True)
print(f"sorted (desc):    {copia}")

# sorted() restituisce una nuova lista senza modificare l'originale
print(f"sorted() nuova:   {sorted(numeri)}")
print(f"Originale intatta: {numeri}")
````

### 1.4 List comprehension — Creare liste in modo compatto

Una **list comprehension** è un modo elegante per creare una lista applicando un'espressione a ogni elemento di un'altra sequenza. È uno dei pattern più idiomatici di Python.

```python
# Forma estesa:
quadrati = []
for n in range(1, 6):
    quadrati.append(n ** 2)

# Equivalente con comprehension:
quadrati = [n ** 2 for n in range(1, 6)]
```

Sintassi: `[espressione  for  variabile  in  sequenza  if  condizione]`

````python
# Quadrati da 1 a 10
quadrati = [n ** 2 for n in range(1, 11)]
print(f"Quadrati:      {quadrati}")

# Solo i pari
pari = [n for n in range(1, 21) if n % 2 == 0]
print(f"Pari 1-20:     {pari}")

# Trasformare string
nomi        = ["ana", "luis", "maria", "carlos"]
maiuscole   = [n.capitalize() for n in nomi]
print(f"Capitalizzati: {maiuscole}")

# Filtrare i voti sufficienti
voti      = [4, 8, 3, 9, 5, 2, 7, 6]
promossi  = [n for n in voti if n >= 5]
print(f"Promossi:      {promossi}")
````

## 2. Tuple (`tuple`)

Una tupla è come una lista, ma **immutabile**: una volta creata, non puoi modificarla. Si scrive con le parentesi tonde.

```python
coordinate  = (40.4168, -3.7038)   # latitudine, longitudine di Madrid
colori_rgb  = (255, 128, 0)        # rosso, verde, blu
vuota       = ()
un_elemento = (42,)                # ← virgola obbligatoria per tuple di 1 elemento
```

### Quando usare una tupla invece di una lista?

| Situazione | Usa |
|-----------|-----|
| I dati **non devono cambiare** (coordinate, configurazione, costanti) | `tuple` |
| I dati **possono cambiare** (collezione dinamica) | `list` |
| Come **chiave di dizionario** (le chiavi devono essere immutabili) | `tuple` |
| Restituire **più valori** da una funzione | `tuple` |

````python
# Creare tuple
coordinate = (40.4168, -3.7038)
rgb_rosso  = (255, 0, 0)

print(coordinate)
print(type(coordinate))    # <class 'tuple'>

# Accesso come per le liste
print(coordinate[0])       # latitudine
print(coordinate[1])       # longitudine

# Spacchettamento (unpacking): assegnare ogni elemento a una variabile
latitudine, longitudine = coordinate
print(f"Latitudine: {latitudine}, Longitudine: {longitudine}")

# Scambiare due variabili con una tupla (pattern elegante di Python)
a, b = 10, 20
a, b = b, a    # swap senza variabile temporanea
print(f"a={a}, b={b}")
````

````python
# Le tuple sono immutabili — provare a modificarle dà errore
punto = (3, 7)

# Questo funziona:
print(punto[0])
print(len(punto))
print(3 in punto)

# Decommenta per vedere l'errore:
# punto[0] = 10   # TypeError: 'tuple' object does not support item assignment
````

````python
# Spacchettamento avanzato con * (cattura il resto)
primo, *resto = (1, 2, 3, 4, 5)
print(f"Primo: {primo}")       # 1
print(f"Resto: {resto}")       # [2, 3, 4, 5]

*inizio, ultimo = (1, 2, 3, 4, 5)
print(f"Inizio: {inizio}")     # [1, 2, 3, 4]
print(f"Ultimo: {ultimo}")     # 5

# Uso tipico: funzione che restituisce più valori
def minmax(lista):
    return min(lista), max(lista)  # restituisce una tupla

minimo, massimo = minmax([4, 7, 2, 9, 1])
print(f"Minimo: {minimo}, Massimo: {massimo}")
````

## 3. Insiemi (`set`)

Un insieme (set) è una collezione **non ordinata** e **senza elementi duplicati**. Si scrive con le parentesi graffe (ma senza chiavi).

```python
colori = {"rosso", "verde", "blu"}
```

I suoi due superpoteri:
1. **Elimina i duplicati automaticamente**
2. **Operazioni di teoria degli insiemi** (unione, intersezione, differenza)

> ⚠️ `{}` crea un dizionario vuoto, non un set. Per un set vuoto usa `set()`.

````python
# Creare set
colori = {"rosso", "verde", "blu", "rosso"}  # il duplicato scompare
print(colori)          # l'ordine può variare
print(type(colori))

# Convertire una lista in set per eliminare i duplicati
lista_con_duplicati = [1, 2, 2, 3, 3, 3, 4]
senza_duplicati = set(lista_con_duplicati)
print(senza_duplicati)

# Riconvertire in lista
lista_pulita = sorted(senza_duplicati)  # sorted() per ordinarla
print(lista_pulita)

# Verificare l'appartenenza — molto efficiente nei set
print("rosso" in colori)   # True
print("nero" in colori)    # False
````

````python
# Operazioni sugli insiemi
matematica = {"Ana", "Luis", "Maria", "Carlos"}
fisica     = {"Luis", "Maria", "Pedro", "Laura"}

# Unione: tutti quelli che sono in almeno uno dei due
print(f"Unione (|):            {matematica | fisica}")

# Intersezione: solo quelli che sono in entrambi
print(f"Intersezione (&):      {matematica & fisica}")

# Differenza: quelli che sono nel primo ma non nel secondo
print(f"Differenza (-):        {matematica - fisica}")

# Differenza simmetrica: quelli che sono nell'uno o nell'altro, ma non in entrambi
print(f"Diff. simmetrica (^):  {matematica ^ fisica}")
````

````python
# Metodi di modifica
tecnologie = {"Python", "SQL", "Git"}

tecnologie.add("Docker")          # aggiunge un elemento
print(f"Dopo add:      {tecnologie}")

tecnologie.update({"AWS", "Linux"})  # aggiunge più elementi insieme
print(f"Dopo update:   {tecnologie}")

tecnologie.discard("Git")         # elimina se esiste (nessun errore se non esiste)
print(f"Dopo discard:  {tecnologie}")

# issubset / issuperset
basilari = {"Python", "SQL"}
print(f"basilari è sottoinsieme di tecnologie? {basilari.issubset(tecnologie)}")
````

## 4. Dizionari (`dict`)

Un dizionario memorizza coppie **chiave → valore**. Invece di accedere per posizione numerica, accedi per **chiave** — qualsiasi valore immutabile (string, numero, tupla).

```python
persona = {
    "nome":   "Ana",
    "eta":    28,
    "citta":  "Madrid"
}
```

```
chiave   →   valore
"nome"   →   "Ana"
"eta"    →   28
"citta"  →   "Madrid"
```

- Le **chiavi** sono uniche — non possono essercene due uguali.
- I **valori** possono essere di qualsiasi tipo, anche altre liste o dizionari.
- Da Python 3.7, i dizionari **mantengono l'ordine di inserimento**.

````python
# Creare un dizionario
persona = {
    "nome":  "Ana",
    "eta":   28,
    "citta": "Madrid"
}

print(persona)
print(type(persona))

# Accesso per chiave
print(persona["nome"])    # Ana
print(persona["eta"])     # 28

# .get() — accesso sicuro: restituisce None (o un default) se la chiave non esiste
print(persona.get("email"))              # None
print(persona.get("email", "nessuna email")) # nessuna email
````

````python
# Modificare i dizionari
persona = {"nome": "Ana", "eta": 28}

# Aggiungere o modificare
persona["citta"] = "Madrid"    # aggiunge una chiave nuova
persona["eta"]   = 29          # modifica una chiave esistente
print(f"Dopo la modifica: {persona}")

# Eliminare
del persona["citta"]
print(f"Dopo del:         {persona}")

valore = persona.pop("eta")     # elimina e restituisce il valore
print(f"Dopo pop:         {persona}  ← eta eliminata: {valore}")

# Unire due dizionari (Python 3.9+: operatore |)
base  = {"nome": "Ana", "eta": 28}
extra = {"citta": "Madrid", "email": "ana@esempio.com"}
completo = base | extra
print(f"Unito:            {completo}")
````

````python
# Iterare sui dizionari
persona = {
    "nome":  "Ana",
    "eta":   28,
    "citta": "Madrid",
    "email": "ana@esempio.com"
}

# Solo chiavi
print("Chiavi:")
for chiave in persona:
    print(f"  {chiave}")

# Solo valori
print("Valori:")
for valore in persona.values():
    print(f"  {valore}")

# Chiavi e valori insieme (il caso più comune)
print("Coppie chiave-valore:")
for chiave, valore in persona.items():
    print(f"  {chiave:10} → {valore}")
````

````python
# Dizionari annidati — molto comuni per rappresentare dati strutturati
rubrica = {
    "ana": {
        "telefono": "600111222",
        "email":    "ana@esempio.com",
        "citta":    "Madrid"
    },
    "luis": {
        "telefono": "600333444",
        "email":    "luis@esempio.com",
        "citta":    "Barcellona"
    }
}

# Accesso annidato
print(rubrica["ana"]["telefono"])   # 600111222

# Iterare
for nome, dati in rubrica.items():
    print(f"{nome.capitalize()}: {dati['telefono']} ({dati['citta']})")
````

### 4.1 Dict comprehension

````python
# Dict comprehension: {chiave: valore for ...}

# Quadrati da 1 a 5
quadrati = {n: n**2 for n in range(1, 6)}
print(quadrati)    # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Invertire un dizionario (valore → chiave)
originale = {"a": 1, "b": 2, "c": 3}
invertito = {v: k for k, v in originale.items()}
print(invertito)    # {1: 'a', 2: 'b', 3: 'c'}

# Filtrare solo i voti sufficienti
voti = {"Ana": 8.5, "Luis": 4.0, "Maria": 7.0, "Carlos": 3.5}
promossi = {studente: voto for studente, voto in voti.items() if voto >= 5}
print(promossi)
````

## Confronto tra le collezioni

| | `list` | `tuple` | `set` | `dict` |
|-|--------|---------|-------|--------|
| Si crea con | `[...]` | `(...)` | `{...}` | `{k: v}` |
| Ordinata | ✅ | ✅ | ❌ | ✅ (3.7+) |
| Mutabile | ✅ | ❌ | ✅ | ✅ |
| Duplicati | ✅ | ✅ | ❌ | ❌ (chiavi) |
| Accesso per | indice | indice | — | chiave |
| Caso d'uso principale | Sequenze dinamiche | Dati fissi | Unicità e teoria degli insiemi | Dati chiave-valore |

```python
def aggiungi_contatto(rubrica, nome, telefono, email):
    chiave = nome.split()[0].lower()   # prima parola del nome come chiave
    if chiave in rubrica:
        print(f"Esiste già un contatto con chiave '{chiave}'.")
        return
    rubrica[chiave] = {"nome": nome, "telefono": telefono, "email": email}
    print(f"Contatto '{nome}' aggiunto.")

def modifica_contatto(rubrica, chiave, campo, nuovo_valore):
    chiave = chiave.lower()
    if chiave not in rubrica:
        print(f"Non esiste nessun contatto con chiave '{chiave}'.")
        return
    campi_validi = {"nome", "telefono", "email"}
    if campo not in campi_validi:
        print(f"Campo non valido. Campi disponibili: {campi_validi}")
        return
    rubrica[chiave][campo] = nuovo_valore
    print(f"Campo '{campo}' aggiornato per {rubrica[chiave]['nome']}.")

def elimina_contatto(rubrica, chiave):
    chiave = chiave.lower()
    if chiave not in rubrica:
        print(f"Non esiste nessun contatto con chiave '{chiave}'.")
        return
    nome = rubrica[chiave]['nome']
    conferma = input(f"Eliminare {nome}? (s/n): ")
    if conferma.lower() == "s":
        del rubrica[chiave]
        print(f"Contatto '{nome}' eliminato.")
    else:
        print("Eliminazione annullata.")

# Prova
aggiungi_contatto(rubrica, "Carlos Ruiz", "600777888", "carlos@mail.com")
modifica_contatto(rubrica, "carlos", "telefono", "611999000")
mostra_rubrica(rubrica)
```

**Punti chiave:**
- `campi_validi = {"nome", "telefono", "email"}` è un **set** — lo usiamo per validare che il campo esista. L'operatore `in` su un set è più efficiente che su una lista.
- `chiave not in rubrica` verifica l'esistenza prima di operare — evita `KeyError`.
- Separare la logica in funzioni permette di richiamare `mostra_rubrica` in qualsiasi punto senza duplicare codice.

```python
        print(f"Contatti con etichetta '{etichetta}': {trovati}")
    else:
        print(f"Nessun contatto ha l'etichetta '{etichetta}'.")

def mostra_etichette(rubrica):
    """Mostra tutte le etichette uniche usate nella rubrica."""
    tutte = set()   # set per accumulare senza duplicati
    for dati in rubrica.values():
        tutte.update(dati["etichette"])
    print(f"Etichette in uso: {sorted(tutte)}")

cerca_per_etichetta(rubrica_con_etichette, "lavoro")
cerca_per_etichetta(rubrica_con_etichette, "amico")
mostra_etichette(rubrica_con_etichette)
```

**Punti chiave:**
- `etichetta in dati["etichette"]` è efficiente perché `dati["etichette"]` è un **set** — la ricerca è quasi istantanea indipendentemente dalla dimensione.
- La list comprehension con `if` filtra sulla stessa riga in cui costruisce la lista.
- `tutte.update(...)` aggiunge tutti gli elementi di un set a un altro — equivale all'unione accumulata.
- `sorted(tutte)` converte il set in una lista ordinata per mostrarla in modo leggibile.

```
Storico di Ana García:
  2025-01-10  →  120 s  (2 min 0 s)
  2025-01-15  →   45 s  (0 min 45 s)
```

```python
def registra_chiamata(rubrica, chiave, data, durata):
    chiave = chiave.lower()
    if chiave not in rubrica:
        print("Contatto non trovato.")
        return
    rubrica[chiave]["chiamate"].append((data, durata))  # tupla (data, durata)
    print(f"Chiamata registrata per {rubrica[chiave]['nome']}.")

def mostra_storico(rubrica, chiave):
    chiave = chiave.lower()
    if chiave not in rubrica:
        print("Contatto non trovato.")
        return
    dati     = rubrica[chiave]
    chiamate = dati["chiamate"]
    print(f"Storico di {dati['nome']}:")
    totale = 0
    for data, secondi in chiamate:            # spacchettamento di tupla
        minuti = secondi // 60
        secs   = secondi % 60
        print(f"  {data}  →  {secondi:4} s  ({minuti} min {secs} s)")
        totale += secondi
    min_totali = totale // 60
    sec_totali = totale % 60
    print(f"Totale: {totale} s ({min_totali} min {sec_totali} s)")

registra_chiamata(rubrica_chiamate, "ana", "2025-01-25", 90)
mostra_storico(rubrica_chiamate, "ana")
```

**Punti chiave:**
- Le chiamate vengono memorizzate come **tuple** perché, una volta registrate, non devono essere modificate — sono dati storici.
- `for data, secondi in chiamate` usa lo **spacchettamento di tupla** direttamente nel `for` — elegante e leggibile.
- La lista `"chiamate"` è mutabile (possiamo fare `.append()`), ma ogni elemento della lista è una tupla immutabile.
