## 1. Il problema: il codice ha bisogno di decidere

Immagina di voler scrivere un programma che dica se una persona può votare. Senza condizionali, non puoi farlo — il codice farebbe la stessa cosa per qualsiasi età:

```python
eta = 15
print("Puoi votare")   # questo viene stampato sempre, anche se l'età è 15
```

Ci serve una struttura che dica: **«se l'età è maggiore o uguale a 18, allora mostra il messaggio»**.

È esattamente quello che fa `if`.

---
## 2. `if` — Se la condizione si verifica, esegui questo blocco

### Sintassi

```python
if condizione:
    # blocco di codice
    # queste righe si eseguono solo se condizione è True
```

**Regole obbligatorie:**
1. La condizione è seguita dai **due punti** `:`
2. Il blocco interno va **indentato** — spostato di 4 spazi (o 1 tabulazione) rispetto all'`if`
3. Python usa l'**indentazione** per sapere quale codice appartiene al blocco — non usa le parentesi graffe `{}` come altri linguaggi

```
if condizione:          ← due punti obbligatori
    istruzione_1        ← indentata (appartiene all'if)
    istruzione_2        ← indentata (appartiene all'if)
istruzione_3             ← NON indentata (si esegue sempre)
```

````python
eta = 20

if eta >= 18:
    print("Puoi votare")            # si esegue solo se eta >= 18
    print("Ricorda di portare la carta d'identità")  # appartiene anch'essa all'if

print("Programma terminato")        # questo si esegue sempre
````

````python
# Prova a cambiare il valore di eta a 15 ed esegui di nuovo
eta = 15

if eta >= 18:
    print("Puoi votare")
    print("Ricorda di portare la carta d'identità")

print("Programma terminato")   # questo compare comunque, anche se eta è 15
````

### ⚠️ L'indentazione è obbligatoria e precisa

L'indentazione scorretta è uno degli errori più comuni in Python. Vediamo i tipi di errore:

````python
# Errore 1: manca l'indentazione → IndentationError
# Decommenta per vedere l'errore:
# if True:
# print("manca l'indentazione")   ← IndentationError

# Errore 2: indentazione incoerente
# if True:
#     print("4 spazi")
#       print("6 spazi")    ← IndentationError

# CORRETTO: 4 spazi coerenti (oppure usa Tab)
x = 10
if x > 5:
    print("x è maggiore di 5")
    print("anche questo è nel blocco")
print("questo è fuori dal blocco")
````

## 3. `else` — Cosa fare quando la condizione NON si verifica

`else` definisce un blocco che si esegue **solo quando la condizione dell'`if` è `False`**. Va sempre abbinato a un `if`.

```python
if condizione:
    # si esegue se condizione è True
else:
    # si esegue se condizione è False
```

Solo uno dei due blocchi viene eseguito — mai entrambi contemporaneamente.

````python
eta = 16

if eta >= 18:
    print("Accesso consentito")
else:
    print("Accesso negato: devi avere almeno 18 anni")

print("---")

# Un altro esempio: numero positivo o negativo
numero = -7

if numero >= 0:
    print(f"{numero} è positivo (o zero)")
else:
    print(f"{numero} è negativo")
````

````python
# Esempio pratico: divisione sicura
a = 10
b = 0

if b != 0:
    risultato = a / b
    print(f"{a} / {b} = {risultato}")
else:
    print("Errore: non è possibile dividere per zero")
````

## 4. `elif` — Più percorsi

Quando ci sono più di due possibilità, usiamo `elif` (abbreviazione di *else if*). Python valuta le condizioni **dall'alto verso il basso** ed esegue solo il primo blocco la cui condizione è `True`. Il resto viene saltato.

```python
if condizione_1:
    # blocco 1
elif condizione_2:
    # blocco 2
elif condizione_3:
    # blocco 3
else:
    # blocco predefinito (nessuna condizione precedente era True)
```

**Punti chiave:**
- Possono esserci **zero o più** `elif` tra l'`if` e l'`else`.
- L'`else` finale è **opzionale** — ma se lo aggiungi, funziona come rete di sicurezza.
- Si esegue **uno solo** di tutti i blocchi.

````python
# Classificazione per voto
voto = 7.5

if voto >= 9:
    valutazione = "Ottimo"
elif voto >= 7:
    valutazione = "Buono"
elif voto >= 5:
    valutazione = "Sufficiente"
else:
    valutazione = "Insufficiente"

print(f"Voto: {voto} → {valutazione}")
````

````python
# ⚠️ L'ORDINE degli elif è importante
# Python si ferma alla prima condizione vera

voto = 9.5

# ORDINE CORRETTO: dal più alto al più basso
if voto >= 9:
    print(f"Corretto → Ottimo ({voto})")
elif voto >= 7:
    print(f"Buono ({voto})")
elif voto >= 5:
    print(f"Sufficiente ({voto})")

print("---")

# ORDINE SCORRETTO: dal più basso al più alto → errore logico silenzioso
if voto >= 5:
    print(f"Scorretto → Sufficiente ({voto})")   # si esegue anche se è 9.5
elif voto >= 7:
    print(f"Buono ({voto})")                     # non ci arriva mai
elif voto >= 9:
    print(f"Ottimo ({voto})")                    # non ci arriva mai
````

````python
# Esempio: classificazione della temperatura
temperatura = float(input("Qual è la temperatura esterna (°C)? "))

if temperatura <= 0:
    descrizione = "gelida 🧊"
    consiglio   = "Copriti bene"
elif temperatura <= 10:
    descrizione = "molto fredda"
    consiglio   = "Indossa un cappotto"
elif temperatura <= 20:
    descrizione = "fresca"
    consiglio   = "Una giacca può andare bene"
elif temperatura <= 28:
    descrizione = "piacevole"
    consiglio   = "Temperatura ideale"
else:
    descrizione = "calda 🌡️"
    consiglio   = "Idratati bene"

print(f"\nTemperatura {descrizione} ({temperatura}°C)")
print(f"Consiglio: {consiglio}")
````

## 5. Annidamento — `if` dentro un `if`

Puoi mettere un `if` dentro il blocco di un altro `if`. Questo si chiama **annidamento** e serve a verificare condizioni più specifiche.

```python
if condizione_esterna:
    # arriviamo qui solo se condizione_esterna è True
    if condizione_interna:
        # arriviamo qui solo se entrambe sono True
    else:
        # condizione_esterna True, condizione_interna False
else:
    # condizione_esterna è False
```

> ⚠️ **Attenzione:** non annidare più di 2-3 livelli. Il codice diventa difficile da leggere. Se hai molto annidamento, probabilmente puoi riscrivere le condizioni con `and`.

````python
# Esempio: accesso a un sistema
utente_attivo = True
password_ok   = True
is_admin      = False

if utente_attivo:
    if password_ok:
        print("Accesso concesso")
        if is_admin:
            print("Benvenuto nel pannello di amministrazione")
        else:
            print("Benvenuto nell'area utente")
    else:
        print("Password errata")
else:
    print("Account disattivato")
````

````python
# La stessa logica, scritta in modo più piatto con 'and'
# Più leggibile quando le condizioni sono indipendenti

utente_attivo = True
password_ok   = True
is_admin      = False

if not utente_attivo:
    print("Account disattivato")
elif not password_ok:
    print("Password errata")
elif is_admin:
    print("Accesso concesso — pannello di amministrazione")
else:
    print("Accesso concesso — area utente")
````

## 6. Buone pratiche

### 6.1 Condizioni leggibili

Evita condizioni inutilmente complicate:

````python
attivo = True

# ❌ Ridondante — confrontare un bool con True è inutile
if attivo == True:
    print("attivo (forma ridondante)")

# ✅ Idiomatico — il bool è già la condizione
if attivo:
    print("attivo (forma corretta)")

# ❌ Ridondante per False
if attivo == False:
    print("non attivo (ridondante)")

# ✅ Usa 'not'
if not attivo:
    print("non attivo (corretto)")
````

````python
# 6.2 Clausole di guardia (guard clauses)
# Invece di annidare, scarta i casi non validi all'inizio

# ❌ Logica molto annidata
def elabora_v1(numero):
    if numero is not None:
        if numero > 0:
            if numero % 2 == 0:
                print(f"{numero} è un intero positivo pari")

# ✅ Clausole di guardia: scarta i casi non validi prima
def elabora_v2(numero):
    if numero is None:
        print("Il numero non può essere None")
        return
    if numero <= 0:
        print("Il numero deve essere positivo")
        return
    if numero % 2 != 0:
        print("Il numero deve essere pari")
        return
    print(f"{numero} è un intero positivo pari")

elabora_v1(8)
elabora_v2(8)
elabora_v2(-3)
elabora_v2(7)
````

````python
# 6.3 match/case — Python 3.10+
# Un'alternativa più pulita all'if/elif quando confronti una variabile
# con valori concreti (come uno switch di altri linguaggi)

giorno = "lunedì"

# Con if/elif
if giorno == "lunedì" or giorno == "martedì" or giorno == "mercoledì" or giorno == "giovedì" or giorno == "venerdì":
    tipo = "feriale"
elif giorno == "sabato" or giorno == "domenica":
    tipo = "weekend"
else:
    tipo = "giorno sconosciuto"
print(f"{giorno}: {tipo}")

# Con match/case (più elegante)
match giorno:
    case "lunedì" | "martedì" | "mercoledì" | "giovedì" | "venerdì":
        tipo = "feriale"
    case "sabato" | "domenica":
        tipo = "weekend"
    case _:            # _ è il caso predefinito (come else)
        tipo = "giorno sconosciuto"

print(f"{giorno}: {tipo}")
````

### Esercizio 4 — Media di più materie

**Consegna:** Scrivi un programma che chieda i voti di **tre materie**, calcoli la media e determini se lo studente:
- Ha **superato tutte** le materie (tutte ≥ 5)
- Ha qualche **insufficienza**
- Ha **superato l'anno** (media ≥ 5 e nessuna insufficienza)

Mostra il voto di ogni materia, la media e il risultato finale.

**Esempio di output atteso:**
```
Voto di Matematica:  6
Voto di Italiano:    8
Voto di Storia:      4

--- Riepilogo ---
Matematica:  6.0  ✓ Superata
Italiano:    8.0  ✓ Superata
Storia:      4.0  ✗ Insufficiente

Risultato: Non supera l'anno (ha insufficienze)
```

````python
# ✏️ IL TUO CODICE QUI
````

<details>
<summary>💡 Vedi soluzione — Esercizio 4</summary>

```python
matematica = float(input("Voto di Matematica: "))
italiano   = float(input("Voto di Italiano:   "))
storia     = float(input("Voto di Storia:     "))

media = (matematica + italiano + storia) / 3

matematica_superata = matematica >= 5
italiano_superato    = italiano   >= 5
storia_superata       = storia     >= 5

tutte_superate = matematica_superata and italiano_superato and storia_superata
supera_anno    = media >= 5 and tutte_superate

# Riepilogo per materia
print("\n--- Riepilogo ---")
print(f"Matematica:  {matematica:.1f}  {'✓ Superata' if matematica_superata else '✗ Insufficiente'}")
print(f"Italiano:    {italiano:.1f}  {'✓ Superata' if italiano_superato  else '✗ Insufficiente'}")
print(f"Storia:      {storia:.1f}  {'✓ Superata' if storia_superata else '✗ Insufficiente'}")

print(f"\nMedia: {media:.1f}")

if supera_anno:
    print("Risultato: Supera l'anno 🎉")
elif tutte_superate:
    print("Risultato: Tutte superate ma la media è insufficiente")
else:
    print("Risultato: Non supera l'anno (ha insufficienze)")
```

**Punti chiave:**
- Calcoliamo prima i booleani individuali e poi li combiniamo — codice più leggibile.
- Il ternario `'✓ Superata' if matematica_superata else '✗ Insufficiente'` dentro la f-string permette un output compatto.
- L'`elif tutte_superate` cattura il caso speciale: tutte superate ma media bassa — un caso limite che merita un messaggio proprio.
</details>

### Esercizio 5 (sfida) — Sistema di borse di studio

**Consegna:** Un'università assegna borse di studio secondo i seguenti criteri:

| Borsa | Condizione |

| **Borsa di eccellenza** | Media ≥ 9 E tutti i compiti consegnati |

Scrivi un programma che chieda:
- La **media dei voti** dello studente
- Se ha qualche **insufficienza** (s/n)
- Se ha consegnato **tutti i compiti** (s/n)

E mostri quale borsa gli spetta con una breve descrizione.

**Esempio di output atteso:**
```
Media dei voti: 9.2
Ha qualche insufficienza? (s/n): n
Ha consegnato tutti i compiti? (s/n): s

--- Assegnazione borsa ---
Media:            9.2
Insufficienze:    No
Compiti:          Tutti consegnati

✅ Borsa di eccellenza
Copre il 100% dell'iscrizione più un aiuto mensile.
```

````python
# ✏️ IL TUO CODICE QUI
````

<details>
<summary>💡 Vedi soluzione — Esercizio 5</summary>

```python
media                 = float(input("Media dei voti: "))
ha_insufficienze      = input("Ha qualche insufficienza? (s/n): ").lower() == "s"

senza_insufficienze = not ha_insufficienze

print(f"Media:            {media}")
print(f"Insufficienze:    {'Sì' if ha_insufficienze else 'No'}")

if media >= 9 and senza_insufficienze and compiti_ok:
    borsa       = "Borsa di eccellenza"
    descrizione = "Copre il 100% dell'iscrizione più un aiuto mensile."
elif media >= 7 and senza_insufficienze:
    borsa       = "Borsa di rendimento"
    descrizione = "Copre il 75% dell'iscrizione."
elif media >= 5 and senza_insufficienze:
    borsa       = "Borsa di continuità"

icona = "✅" if borsa != "Nessuna borsa" else "❌"

print(descrizione)
```

**Punti chiave:**
- `senza_insufficienze = not ha_insufficienze` crea una variabile con nome positivo — più leggibile che usare `not` direttamente nella condizione.
- Gli `if/elif` vanno dalla borsa più esigente alla meno esigente — se una condizione non si verifica, si prova la successiva con requisiti più bassi.
- L'`else` funziona come rete di sicurezza: cattura qualsiasi caso che non rientri nelle borse precedenti.
- Separare il calcolo (`if/elif/else`) dalla presentazione (`print`) rende il codice più organizzato e facile da modificare.
</details>

## Riepilogo della lezione

### Struttura generale

```python
if condizione_1:          # obbligatorio, sempre per primo
    blocco_1
elif condizione_2:        # opzionale, può ripetersi
    blocco_2
elif condizione_3:
    blocco_3
else:                     # opzionale, sempre alla fine
    blocco_predefinito
```

### Regole essenziali

| Regola | Dettaglio |
|-------|---------|
| Due punti `:` | Obbligatori dopo `if`, `elif` ed `else` |
| Indentazione | 4 spazi — definisce cosa appartiene al blocco |
| Si esegue un solo blocco | Python si ferma alla prima condizione vera |
| Ordine delle condizioni | Dalla più restrittiva alla più generale |
| `else` | Opzionale — cattura il caso predefinito |

### Pattern comuni

```python
# Validazione prima di tutto
if input_non_valido:
    print("Errore")
else:
    # logica principale

# Booleano diretto (senza confrontare con True/False)
if attivo:          # no: if attivo == True
if not attivo:       # no: if attivo == False

# Ternario per valori semplici
etichetta = "pari" if n % 2 == 0 else "dispari"
```
