## 1. El problema sin Git

Antes de ver qué es Git, veamos el problema que resuelve. Te resultará familiar.

### La carpeta del caos

```
📁 mi_proyecto/
├── analisis.py
├── analisis_v2.py
├── analisis_v2_final.py
├── analisis_v2_final_ESTE.py
├── analisis_v2_final_ESTE_de_verdad.py
├── analisis_BUENO.py
├── analisis_BUENO_copia.py
└── analisis_para_entregar_hoy.py
```

¿Cuál es la versión correcta? ¿Qué cambió entre `_v2` y `_v2_final`? ¿Puedes volver a la versión de hace tres días que sí funcionaba?

### El problema en un equipo

```
Lunes:  Ana trabaja en analisis.py  →  sube analisis_ANA.py
Martes: Luis trabaja en analisis.py →  sube analisis_LUIS.py
Miércoles: ¿Quién fusiona los dos? ¿Qué pasa con los cambios que se pisan?
```

### Los cinco problemas que Git resuelve

| Problema | Sin Git | Con Git |
|----------|---------|--------|
| **Historial** | Nombres de archivo | Cada cambio registrado con fecha, autor y mensaje |
| **Deshacer** | Imposible o manual | `git revert` — vuelve a cualquier punto anterior |
| **Colaboración** | Copias que se pisan | Cada persona trabaja en su rama, Git fusiona |
| **Contexto** | ¿Por qué cambié esto? | El mensaje de commit explica el motivo |
| **Seguridad** | Un solo punto de fallo | Cada desarrollador tiene una copia completa |

## 2. Qué es Git

**Git** es un sistema de control de versiones distribuido (**DVCS** — Distributed Version Control System). Lo creó Linus Torvalds en 2005 para gestionar el desarrollo del kernel de Linux.

### Cómo funciona Git por dentro

Git no guarda diferencias línea a línea — guarda **instantáneas** (snapshots) del estado completo del proyecto en cada commit.

```
Commit 1      Commit 2      Commit 3      Commit 4
──────────    ──────────    ──────────    ──────────
analisis.py   analisis.py*  analisis.py   analisis.py*
datos.csv     datos.csv     datos.csv*    datos.csv*
config.json   config.json   config.json   config.json*

* = archivo modificado en este commit
  = mismo archivo que el commit anterior (Git lo referencia, no lo duplica)
```

Esto hace Git extremadamente eficiente: no copia archivos que no han cambiado.

### Distribuido vs centralizado

```
Sistema CENTRALIZADO (SVN, antes)   Sistema DISTRIBUIDO (Git)
────────────────────────────────    ─────────────────────────────

       SERVIDOR CENTRAL                   REPOSITORIO REMOTO
           │                                   (GitHub)
    ┌──────┼──────┐                   ┌─────────┼─────────┐
   Ana    Luis   Marta               Ana       Luis      Marta
  (copia (copia (copia              (copia    (copia    (copia
parcial) parcial) parcial)          COMPLETA) COMPLETA) COMPLETA)

Si el servidor cae → trabajo perdido   Si GitHub cae → seguimos trabajando
Sin conexión → sin trabajo             Sin conexión → seguimos con Git local
```

### Vocabulario fundamental

| Término | Qué es |
|---------|--------|
| **Repositorio** (`repo`) | El proyecto completo + todo su historial de cambios |
| **Commit** | Una instantánea guardada del proyecto en un momento dado |
| **Branch** (rama) | Una línea de desarrollo independiente |
| **Merge** | Fusionar dos ramas |
| **Remote** | Una copia del repo en un servidor (GitHub, GitLab…) |
| **Clone** | Descargar un repo remoto completo a tu máquina |
| **Push / Pull** | Enviar / recibir cambios al/del remoto |

### Dónde se guarda la configuración

Git tiene tres niveles de configuración — cada uno sobreescribe al anterior:

```
SISTEMA         →  /etc/gitconfig           (afecta a todos los usuarios del SO)
GLOBAL          →  ~/.gitconfig             (afecta a todos tus proyectos)
LOCAL           →  mi_proyecto/.git/config  (afecta solo a ese proyecto)

Prioridad: LOCAL > GLOBAL > SISTEMA
```

Puedes tener un email personal en `--global` y un email de empresa en `--local` dentro de un proyecto concreto.

## 4. Las tres zonas de Git

Este es el concepto más importante para entender cómo funciona Git. Todo archivo en tu proyecto puede estar en una de tres zonas:

```
┌─────────────────────────────────────────────────────────────┐
│                      TU PROYECTO                           │
│                                                             │
│  WORKING DIRECTORY    STAGING AREA       REPOSITORIO       │
│  (directorio de        (índice /           (.git/)          │
│   trabajo)             área de stage)                       │
│                                                             │
│  Los archivos que      Lo que vas a        El historial     │
│  editas con tu         incluir en el       permanente de    │
│  editor                próximo commit      commits          │
│                                                             │
│       │                    │                    │           │
│       │   git add          │   git commit       │           │
│       │ ─────────────────► │ ─────────────────► │           │
│       │                    │                    │           │
│       │ ◄─────────────────────────────────────  │           │
│       │   git checkout / git restore            │           │
└─────────────────────────────────────────────────────────────┘
```

### ¿Por qué existe el Staging Area?

Puede parecer un paso extra innecesario. Pero es una de las ideas más brillantes de Git:

```
Situación sin staging area:
  Hice 5 cambios en 5 archivos → solo puedo hacer UN commit con todo

Situación con staging area:
  Hice 5 cambios en 5 archivos
  → Añado al stage los 2 archivos del bug que arreglé → commit 1: "fix: corregir cálculo de IVA"
  → Añado al stage los 3 archivos de la nueva función → commit 2: "feat: añadir exportación a PDF"
  → Historial limpio, legible, con propósito
```

### Estados de un archivo

```
Untracked    →  Git no sabe que existe (archivo nuevo)
Tracked      →  Git lo conoce. Puede estar en:
  Unmodified →  igual que el último commit
  Modified   →  cambiado pero no añadido al stage
  Staged     →  preparado para el próximo commit
```

## 6. `.gitignore` — Qué debe ignorar Git

Antes de hacer el primer commit, configuramos qué archivos **nunca** deben entrar al repositorio. El archivo `.gitignore` contiene patrones de nombres de archivo o directorio que Git ignorará completamente.

**¿Qué se ignora habitualmente?**

| Categoría | Ejemplos | Por qué |
|-----------|----------|--------|
| **Secretos** | `.env`, `*.key`, `secretos.*` | Nunca subir credenciales o claves API |
| **Entornos virtuales** | `venv/`, `.venv/`, `env/` | Pesados y se regeneran con `pip install` |
| **Caché de Python** | `__pycache__/`, `*.pyc` | Archivos compilados, se generan solos |
| **Datos grandes** | `*.csv`, `*.parquet`, `data/raw/` | Usar Git LFS o almacenamiento externo |
| **Archivos del editor** | `.vscode/`, `.idea/`, `*.swp` | Configuración local que no es del proyecto |
| **Builds y distribución** | `dist/`, `build/`, `*.egg-info/` | Se generan con el proceso de build |

````python
# Crear el .gitignore antes del primer commit
gitignore_content = """# ─── Secretos y configuración local ───────────────────────
.env
*.env
secretos.*
*.key
*.pem

# ─── Python ────────────────────────────────────────────────
__pycache__/
*.py[cod]
*.pyo
*.pyd

# ─── Entornos virtuales ────────────────────────────────────
venv/
.venv/
env/
.env/

# ─── Jupyter ───────────────────────────────────────────────
.ipynb_checkpoints/

# ─── Editores ──────────────────────────────────────────────
.vscode/
.idea/
*.swp
*.swo
.DS_Store

# ─── Datos (grandes o sensibles) ───────────────────────────
data/raw/
*.parquet
"""

Path(".gitignore").write_text(gitignore_content, encoding="utf-8")
print(".gitignore creado")
print()

# Ahora git status no muestra secretos.env — está ignorado
!git status
````

````python
# Verificar explícitamente que secretos.env está ignorado
!git check-ignore -v secretos.env
````

### `git commit` — Guardar la instantánea

Un commit es permanente e inmutable. Una vez hecho, ese momento del proyecto queda registrado para siempre en el historial.

**Anatomía de un buen mensaje de commit:**

```
tipo(ámbito): descripción corta en imperativo (max 72 caracteres)

Cuerpo opcional: explica el POR QUÉ, no el QUÉ.
El código ya muestra el qué.

Closes #123
```

**Tipos más usados (Conventional Commits):**

| Tipo | Cuándo usarlo |
|------|---------------|
| `feat` | Nueva funcionalidad |
| `fix` | Corrección de un bug |
| `docs` | Solo documentación |
| `refactor` | Código reorganizado sin cambiar comportamiento |
| `test` | Añadir o corregir tests |
| `chore` | Mantenimiento: dependencias, configuración |
| `init` | Commit inicial del repositorio |

## Referencia de comandos de esta clase

```bash
# ─── Configuración (una sola vez) ─────────────────────────
git config --global user.name  "Tu Nombre"
git config --global user.email "tu@email.com"
git config --global init.defaultBranch main
git config --global --list          # ver toda la configuración

# ─── Crear un repositorio ──────────────────────────────────
git init                            # inicializar en el directorio actual
git init nombre-proyecto            # crear directorio nuevo e inicializar

# ─── El ciclo diario ───────────────────────────────────────
git status                          # ver el estado actual (úsalo siempre)
git diff                            # cambios en working directory (no staged)
git diff --staged                   # cambios en staging area (lo que va al commit)

git add archivo.py                  # añadir un archivo al stage
git add carpeta/                    # añadir una carpeta entera
git add .                           # añadir todo lo no ignorado
git add -p                          # añadir trozo a trozo (interactivo)

git commit -m "tipo: mensaje"       # commitear con mensaje inline
git commit                          # abre el editor para el mensaje

# ─── Ver el historial ──────────────────────────────────────
git log                             # historial completo
git log --oneline                   # una línea por commit
git log --oneline --graph --all     # con gráfico de ramas
git log --author="Nombre"           # filtrar por autor
git show <hash>                     # ver los detalles de un commit

# ─── Deshacer ─────────────────────────────────────────────
git restore archivo.py              # descartar cambios en working directory
git restore --staged archivo.py     # sacar del stage sin perder cambios
git revert <hash>                   # crear commit que deshace otro (seguro)
```

from datetime import date

# ── COMMIT 2 ──────────────────────────────────────────────
Path("nota.py").write_text("""
from datetime import date

class Nota:
    _contador = 0

    def __init__(self, titulo, contenido):
        Nota._contador += 1
        self.id              = Nota._contador
        self.titulo          = titulo.strip()
        self.contenido       = contenido.strip()
        self.fecha_creacion  = date.today()

    def __str__(self):
        return f"[{self.id}] {self.titulo} ({self.fecha_creacion})"
""".strip(), encoding="utf-8")

from nota import Nota

class GestorNotas:
    def __init__(self):
        self._notas = []

    def crear(self, titulo, contenido):
        nota = Nota(titulo, contenido)
        self._notas.append(nota)
        return nota

    def listar(self):
        if not self._notas:
            print('No hay notas.')
            return
        for nota in self._notas:
            print(nota)
""".strip(), encoding="utf-8")

    def buscar(self, termino):

        return [

- El commit de `docs` solo modifica el README — no mezcla código y documentación.
</details>

<details>
<summary>💡 Ver solución — Ejercicio 5</summary>

```python
import subprocess
from pathlib import Path

def git(comando, cwd):
    """Ejecuta un comando git y devuelve la salida como lista de líneas."""
    resultado = subprocess.run(
        ["git"] + comando.split(),
        cwd=cwd,
        capture_output=True,
        text=True
    )
    return resultado.stdout.strip()


def auditar_repo(ruta_repo):
    ruta  = Path(ruta_repo).expanduser().resolve()
    nombre = ruta.name

    print(f"\n{'═'*50}")
    print(f"  AUDITORÍA — {nombre}")
    print(f"  {ruta}")
    print(f"{'═'*50}")

    # ── 1. Número de commits ───────────────────────────────
    total_commits = git("rev-list --count HEAD", ruta)
    print(f"\nCommits totales: {total_commits}")

    # ── 2. Historial ──────────────────────────────────────
    print("\nHISTORIAL:")
    # Formato: hash|fecha|autor|mensaje
    log_raw = git('log --pretty=format:%h|%ad|%an|%s --date=short', ruta)
    if log_raw:
        for linea in log_raw.splitlines():
            partes = linea.split("|", 3)
            if len(partes) == 4:
                hash_, fecha, autor, mensaje = partes
                print(f"  {hash_:<8} {fecha}  {autor:<15} {mensaje}")

    # ── 3. Archivos trackeados ────────────────────────────
    print("\nARCHIVOS TRACKEADOS:")
    archivos_raw = git("ls-files", ruta)
    archivos = archivos_raw.splitlines() if archivos_raw else []
    total_bytes = 0
    for nombre_archivo in archivos:
        ruta_archivo = ruta / nombre_archivo
        if ruta_archivo.exists():
            tam = ruta_archivo.stat().st_size
            total_bytes += tam
            tam_str = f"{tam / 1024:.1f} KB" if tam > 1024 else f"{tam} B"
            print(f"  {nombre_archivo:<25} {tam_str:>8}")
    print(f"  {'─'*35}")
    print(f"  {'TOTAL':<25} {total_bytes / 1024:.1f} KB ({len(archivos)} archivos)")

    # ── 4. Archivos ignorados presentes ───────────────────
    print("\nARCHIVOS IGNORADOS (presentes en disco):")
    ignorados_raw = git("ls-files --others --ignored --exclude-standard", ruta)
    ignorados = ignorados_raw.splitlines() if ignorados_raw else []
    if ignorados:
        for f in ignorados:
            print(f"  🚫 {f}")
    else:
        print("  (ninguno)")

    # ── 5. Estado del working tree ────────────────────────
    print("\nESTADO:")
    status = git("status --porcelain", ruta)
    if not status:
        print("  ✅ Working tree limpio — todo commiteado")
    else:
        print("  ⚠️  Hay cambios sin commitear:")
        for linea in status.splitlines():
            estado_codigo = linea[:2].strip()
            archivo = linea[3:]
            significado = {
                "M":  "Modificado",
                "A":  "Nuevo en stage",
                "D":  "Eliminado",
                "??": "Sin trackear",
            }.get(estado_codigo, estado_codigo)
            print(f"  {significado:<20} {archivo}")

    print(f"{'═'*50}")


# Ejecutar la auditoría
auditar_repo("~/sistema_notas")
```

**Puntos clave:**
- `subprocess.run` con `capture_output=True, text=True` es la forma correcta de capturar la salida de un comando shell en Python.
- `git log --pretty=format:...` permite diseñar exactamente qué campos queremos — sin parsear texto libre.
- `git ls-files` lista solo los archivos trackeados. `git ls-files --others --ignored --exclude-standard` lista los ignorados presentes.
- `git status --porcelain` devuelve el estado en formato legible por máquina — fácil de parsear, estable entre versiones de Git.
</details>

## Resumen de la clase

### El ciclo de trabajo fundamental

```
1. Editar archivos en el working directory
       │
       ▼
2. git status          → ver qué ha cambiado
       │
       ▼
3. git add <archivos>  → mover cambios relevantes al stage
       │
       ▼
4. git diff --staged   → revisar exactamente qué va a entrar
       │
       ▼
5. git commit -m "tipo: descripción clara"
       │
       └──► volver al paso 1
```

### Comandos que debes conocer de memoria

```bash
git init                     # crear repo
git status                   # brújula — úsalo siempre
git add .                    # añadir todo al stage
git commit -m "mensaje"      # guardar instantánea
git log --oneline            # ver historial
git diff                     # ver cambios sin añadir
git diff --staged            # ver cambios en el stage
```

### Buenas prácticas desde el día uno

| Práctica | Por qué |
|----------|---------|
| Commits pequeños y frecuentes | Más fácil de entender y deshacer |
| Un commit = un propósito | El historial cuenta una historia |
| `.gitignore` antes del primer commit | Nada sensible entra por accidente |
| `git status` antes de `git commit` | Revisas qué estás guardando |
| Mensajes en imperativo | Convención universal: `"añadir"`, no `"añadido"` |
| Nunca commitear credenciales | Una vez en el historial, es muy difícil borrar |
