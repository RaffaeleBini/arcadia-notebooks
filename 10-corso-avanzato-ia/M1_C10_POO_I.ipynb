## 1. ¿Por qué POO?

Imagina que tienes que gestionar 100 empleados. Sin POO:

```python
# Sin POO — datos sueltos, difícil de mantener
nombre_1  = "Ana"
salario_1 = 2500
nombre_2  = "Luis"
salario_2 = 3000

def calcular_bonus(salario):
    return salario * 0.10
```

Con POO:

```python
# Con POO — datos y comportamiento juntos, fácil de escalar
class Empleado:
    def calcular_bonus(self):
        return self.salario * 0.10

ana  = Empleado("Ana",  2500)
luis = Empleado("Luis", 3000)
ana.calcular_bonus()    # 250
```

## 2. Clases y objetos

Una **clase** es la plantilla — define la estructura.  
Un **objeto** (o instancia) es cada copia concreta creada a partir de esa plantilla.

```
Clase Empleado
  ├── atributos: nombre, salario, departamento
  └── métodos:  saludar(), calcular_bonus()
        │
        ├── Objeto: ana  (nombre="Ana",  salario=2500)
        ├── Objeto: luis (nombre="Luis", salario=3000)
        └── Objeto: marta(nombre="Marta",salario=2800)
```

Analogía: la clase es el **molde** para hacer galletas; cada galleta es un objeto distinto pero todas tienen la misma forma.

````python
# La clase más simple posible
class Perro:
    pass   # 'pass' porque aún no añadimos nada

# Crear objetos (instanciar la clase)
rex   = Perro()
laika = Perro()

print(type(rex))          # <class '__main__.Perro'>
print(isinstance(rex, Perro))  # True
print(rex is laika)       # False — son objetos distintos

# Podemos añadir atributos «al vuelo» (no es la forma correcta, pero muestra el concepto)
rex.nombre = "Rex"
rex.edad   = 3
print(f"{rex.nombre} tiene {rex.edad} años")
````

## 3. El método `__init__` — Constructor

`__init__` es el método que Python llama automáticamente cuando creas un objeto con `Clase()`. Su función es inicializar los atributos del objeto.

```python
class MiClase:
    def __init__(self, param1, param2):
        self.atributo1 = param1
        self.atributo2 = param2
```

**¿Qué es `self`?**

`self` es una referencia al **propio objeto** que se está creando. Python lo pasa automáticamente — tú nunca lo pasas al llamar al método. La convención de llamarlo `self` no es obligatoria pero es universal.

```
Empleado("Ana", 2500)
      │
      └──► Python llama __init__(self=<nuevo objeto>, nombre="Ana", salario=2500)
```

````python
class Perro:
    def __init__(self, nombre, raza, edad):
        self.nombre = nombre
        self.raza   = raza
        self.edad   = edad

# Al instanciar, Python llama __init__ automáticamente
rex   = Perro("Rex",   "Pastor Alemán", 3)
laika = Perro("Laika", "Husky",         5)

print(f"{rex.nombre}   — {rex.raza},   {rex.edad} años")
print(f"{laika.nombre} — {laika.raza}, {laika.edad} años")

# Cada objeto tiene sus propios datos
rex.edad = 4   # modificar un atributo
print(f"\nEdad de Rex actualizada:   {rex.edad}")
print(f"Edad de Laika sin cambios: {laika.edad}")
````

````python
# Parámetros con valores por defecto en __init__
class Perro:
    def __init__(self, nombre, raza="Desconocida", edad=0):
        self.nombre = nombre
        self.raza   = raza
        self.edad   = edad

    def descripcion(self):
        return f"{self.nombre} ({self.raza}, {self.edad} años)"

perro1 = Perro("Max")
perro2 = Perro("Nala", "Labrador")
perro3 = Perro("Thor", "Rottweiler", 2)

for p in [perro1, perro2, perro3]:
    print(p.descripcion())
````

## 4. Atributos

Python tiene dos tipos principales de atributos:

| Tipo | Dónde se define | A quién pertenece | Acceso |
|------|-----------------|-------------------|--------|
| **De instancia** | Dentro de `__init__` con `self.x` | A cada objeto | `objeto.x` |
| **De clase** | Dentro de la clase, fuera de métodos | A la clase (compartido) | `Clase.x` o `objeto.x` |

````python
class Contador:
    total_instancias = 0   # ← atributo de CLASE (compartido)

    def __init__(self, nombre):
        self.nombre = nombre               # ← atributo de INSTANCIA
        Contador.total_instancias += 1     # actualizar el contador compartido

c1 = Contador("primero")
c2 = Contador("segundo")
c3 = Contador("tercero")

print(f"Instancias creadas: {Contador.total_instancias}")  # 3 (atributo de clase)
print(f"Nombre de c1: {c1.nombre}")                        # primero (atributo de instancia)
print(f"Nombre de c2: {c2.nombre}")                        # segundo

# El atributo de clase es compartido — si cambia, cambia para todos
print(f"\nDesde la instancia c1: {c1.total_instancias}")   # 3
print(f"Desde la instancia c2: {c2.total_instancias}")     # 3
````

````python
# ⚠️ Cuidado: si asignas sobre la instancia, creas un atributo nuevo en esa instancia
# y el de clase queda intacto

class Config:
    version = "1.0"   # de clase

cfg1 = Config()
cfg2 = Config()

cfg1.version = "2.0"   # crea un atributo de INSTANCIA en cfg1

print(f"cfg1.version:     {cfg1.version}")    # 2.0 (instancia)
print(f"cfg2.version:     {cfg2.version}")    # 1.0 (clase)
print(f"Config.version:   {Config.version}")  # 1.0 (clase, intacta)

# Verificar si el atributo es de instancia o de clase
print(f"\n'version' en cfg1.__dict__: {'version' in cfg1.__dict__}")  # True
print(f"'version' en cfg2.__dict__: {'version' in cfg2.__dict__}")  # False
````

## 5. Métodos

Un **método** es una función definida dentro de una clase. Hay tres tipos:

| Tipo | Decorador | Primer parámetro | Cuándo usarlo |
|------|-----------|-----------------|---------------|
| De instancia | (ninguno) | `self` | Accede o modifica datos del objeto |
| De clase | `@classmethod` | `cls` | Trabaja con atributos de clase; constructores alternativos |
| Estático | `@staticmethod` | (ninguno) | Utilidad relacionada con la clase pero sin necesitar `self` ni `cls` |

````python
class Rectangulo:
    # ─── Atributo de clase ───────────────────────────────
    unidad = "cm"

    # ─── Constructor ─────────────────────────────────────
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto  = alto

    # ─── Métodos de instancia ────────────────────────────
    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

    def es_cuadrado(self):
        return self.ancho == self.alto

    def escalar(self, factor):
        """Modifica el objeto en su lugar."""
        self.ancho *= factor
        self.alto  *= factor

    # ─── Método de clase (constructor alternativo) ───────
    @classmethod
    def cuadrado(cls, lado):
        """Crea un rectángulo cuadrado sin repetir lógica."""
        return cls(lado, lado)

    @classmethod
    def cambiar_unidad(cls, nueva_unidad):
        cls.unidad = nueva_unidad

    # ─── Método estático ─────────────────────────────────
    @staticmethod
    def es_valido(ancho, alto):
        """Comprueba que las medidas sean positivas — no necesita self ni cls."""
        return ancho > 0 and alto > 0


# Uso
r = Rectangulo(6, 4)
print(f"Área:        {r.area()} {Rectangulo.unidad}²")
print(f"Perímetro:   {r.perimetro()} {Rectangulo.unidad}")
print(f"¿Cuadrado?   {r.es_cuadrado()}")

q = Rectangulo.cuadrado(5)          # constructor alternativo
print(f"\nCuadrado:    {q.ancho}x{q.alto}, área={q.area()}")
print(f"¿Cuadrado?   {q.es_cuadrado()}")

print(f"\n¿0x4 válido? {Rectangulo.es_valido(0, 4)}")   # método estático
print(f"¿5x3 válido? {Rectangulo.es_valido(5, 3)}")

r.escalar(2)
print(f"\nTras escalar x2: {r.ancho}x{r.alto}, área={r.area()}")
````

## 6. Métodos especiales (`dunder`)

Los métodos con doble guion bajo al principio y al final (`__método__`) se llaman **dunder** (double underscore). Python los llama automáticamente en respuesta a operadores o funciones built-in.

| Método | Se llama cuando… | Ejemplo |
|--------|------------------|---------|
| `__init__` | Creas un objeto | `Clase()` |
| `__str__` | Conviertes a string legible | `str(obj)`, `print(obj)` |
| `__repr__` | Representación para debugging | `repr(obj)`, en el REPL |
| `__eq__` | Comparas con `==` | `obj1 == obj2` |
| `__lt__` | Comparas con `<` | `obj1 < obj2` |
| `__len__` | Llamas a `len()` | `len(obj)` |
| `__add__` | Usas `+` | `obj1 + obj2` |
| `__contains__` | Usas `in` | `x in obj` |

````python
class Libro:
    def __init__(self, titulo, autor, paginas):
        self.titulo  = titulo
        self.autor   = autor
        self.paginas = paginas

    def __str__(self):
        """Representación legible para el usuario."""
        return f"«{self.titulo}» de {self.autor} ({self.paginas} págs.)"

    def __repr__(self):
        """Representación técnica para debugging."""
        return f"Libro(titulo={self.titulo!r}, autor={self.autor!r}, paginas={self.paginas})"

    def __eq__(self, otro):
        """Dos libros son iguales si tienen el mismo título y autor."""
        if not isinstance(otro, Libro):
            return NotImplemented
        return self.titulo == otro.titulo and self.autor == otro.autor

    def __lt__(self, otro):
        """Ordenar por número de páginas."""
        return self.paginas < otro.paginas

    def __len__(self):
        return self.paginas


b1 = Libro("1984",          "George Orwell",  328)
b2 = Libro("Dune",          "Frank Herbert",  604)
b3 = Libro("1984",          "George Orwell",  328)  # igual que b1

print(str(b1))          # __str__
print(repr(b2))         # __repr__
print(f"b1 == b3: {b1 == b3}")  # __eq__   → True
print(f"b1 == b2: {b1 == b2}")  # __eq__   → False
print(f"b1 < b2:  {b1 < b2}")   # __lt__   → True (328 < 604)
print(f"len(b2):  {len(b2)}")   # __len__

# sorted() usa __lt__ automáticamente
biblioteca = [b2, b1, Libro("El Principito", "Saint-Exupéry", 96)]
for libro in sorted(biblioteca):
    print(f"  {len(libro):>4} págs — {libro}")
````

## 7. Encapsulación — Atributos privados y `@property`

La encapsulación controla qué partes del objeto son accesibles desde fuera. Python usa convenciones de nomenclatura:

| Convención | Ejemplo | Significado |
|------------|---------|-------------|
| Sin prefijo | `self.nombre` | Público — accesible desde cualquier lugar |
| Un guion bajo | `self._nombre` | Protegido — convención «no toques esto desde fuera» |
| Dos guiones bajos | `self.__nombre` | Privado — Python aplica *name mangling* |

> Python no impide el acceso técnicamente, pero los prefijos son señales claras para otros programadores (y para ti mismo en el futuro).

````python
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular  = titular           # público
        self._saldo   = saldo_inicial     # protegido (convención)
        self.__pin    = "0000"            # privado (name mangling)

    # @property convierte un método en un atributo de solo lectura
    @property
    def saldo(self):
        return self._saldo

    # @saldo.setter permite asignar con validación
    @saldo.setter
    def saldo(self, valor):
        if valor < 0:
            raise ValueError("El saldo no puede ser negativo")
        self._saldo = valor

    def depositar(self, cantidad):
        if cantidad <= 0:
            raise ValueError("La cantidad debe ser positiva")
        self._saldo += cantidad
        return self._saldo

    def retirar(self, cantidad):
        if cantidad > self._saldo:
            raise ValueError("Fondos insuficientes")
        self._saldo -= cantidad
        return self._saldo

    def __str__(self):
        return f"Cuenta de {self.titular}: {self._saldo:.2f} €"


cuenta = CuentaBancaria("Ana", 1000)
print(cuenta)

cuenta.depositar(500)
print(f"Tras depósito:  {cuenta.saldo:.2f} €")   # via @property

cuenta.retirar(200)
print(f"Tras retirada:  {cuenta.saldo:.2f} €")

# El setter valida
try:
    cuenta.saldo = -100
except ValueError as e:
    print(f"Error: {e}")

# El atributo privado __pin aplica name mangling
try:
    print(cuenta.__pin)    # AttributeError
except AttributeError:
    print("No se puede acceder a __pin directamente")
    print(f"(existe como: {cuenta._CuentaBancaria__pin})")  # name mangling
````

````python
# @property para atributos calculados (sin setter — de solo lectura)
import math

class Circulo:
    def __init__(self, radio):
        self.radio = radio

    @property
    def area(self):
        return math.pi * self.radio ** 2

    @property
    def perimetro(self):
        return 2 * math.pi * self.radio

    @radio.setter
    def radio(self, valor):
        if valor <= 0:
            raise ValueError("El radio debe ser positivo")
        self._radio = valor

    @property
    def radio(self):
        return self._radio

    def __str__(self):
        return f"Círculo(r={self.radio}, área={self.area:.2f}, per={self.perimetro:.2f})"

c = Circulo(5)
print(c)
c.radio = 10
print(c)

# area y perimetro se recalculan automáticamente al cambiar el radio
try:
    c.area = 100   # TypeError — no tiene setter
except AttributeError as e:
    print(f"No se puede asignar directamente: {e}")
````

class Empleado:
    # ─── Atributos de clase ───────────────────────────────
    empresa       = "MiEmpresa S.L."
    _contador_ids = 0

    # ─── Constructor ─────────────────────────────────────
    def __init__(self, nombre, apellido, departamento, salario):
        Empleado._contador_ids += 1
        self.id           = f"E{Empleado._contador_ids:03d}"
        self.nombre       = nombre.strip().title()
        self.apellido     = apellido.strip().title()
        self.departamento = departamento
        self.salario      = salario          # usa el setter con validación

    # ─── Propiedad salario ───────────────────────────────
    @property
    def salario(self):
        return self._salario

    @salario.setter
    def salario(self, valor):
        if not isinstance(valor, (int, float)) or valor < 0:
            raise ValueError(f"Salario inválido: {valor}. Debe ser un número ≥ 0")
        self._salario = float(valor)

    # ─── Métodos dunder ──────────────────────────────────
    def __str__(self):
        return (
            f"[{self.id}] {self.nombre} {self.apellido} "
            f"— {self.departamento} ({self._salario:,.2f} €)"
        )

    def __repr__(self):
        return (
            f"Empleado(id={self.id!r}, nombre={self.nombre!r}, "
            f"apellido={self.apellido!r}, departamento={self.departamento!r}, "
            f"salario={self._salario})"
        )

    def __eq__(self, otro):
        if not isinstance(otro, Empleado):
            return NotImplemented
        return self.id == otro.id

**Puntos clave:**
- `Empleado._contador_ids += 1` dentro de `__init__` garantiza que el id se asigna antes de cualquier otra operación.
- `self.salario = salario` (sin guion bajo) en el `__init__` pasa por el setter — la validación se aplica ya en la creación.
- `.title()` normaliza el nombre sin importar cómo lo escriba el usuario.
- `__eq__` devuelve `NotImplemented` (no `False`) cuando el otro objeto no es un `Empleado` — así Python puede intentar la comparación inversa.

    @classmethod
    def total_empleados(cls):
        return cls._contador_ids

    @classmethod
    def crear_desde_dict(cls, datos):
        """Constructor alternativo desde un diccionario."""
        return cls(
            datos["nombre"],
            datos["apellido"],
            datos["departamento"],
            datos["salario"]
        )

    @classmethod
    def crear_desde_csv_linea(cls, linea):
        """Constructor alternativo desde una línea CSV."""
        partes = [p.strip() for p in linea.split(",")]
        if len(partes) != 4:
            raise ValueError(f"Línea CSV inválida: se esperaban 4 campos, llegaron {len(partes)}")
        nombre, apellido, departamento, salario_str = partes
        try:
            salario = float(salario_str)
        except ValueError:
            raise ValueError(f"El salario '{salario_str}' no es un número válido")
        return cls(nombre, apellido, departamento, salario)

    @staticmethod
    def es_salario_valido(salario):
        """Devuelve True si el salario es un número >= 0."""
        return isinstance(salario, (int, float)) and salario >= 0

    @staticmethod
    def formatear_nombre(nombre):
        """Capitaliza correctamente nombres compuestos."""
        return " ".join(parte.capitalize() for parte in nombre.split())

### Ejercicio 4 — La clase `Departamento`

**Enunciado:** Crea una clase `Departamento` que gestione una colección de empleados:

- **`__init__(nombre)`** — inicializa con nombre y lista vacía de empleados.
- **`añadir(empleado)`** — añade un `Empleado`. Lanza `TypeError` si no es instancia de `Empleado`.

- **`buscar(id_empleado)`** — devuelve el empleado o `None`.

- **`el_mas_pagado()`** — devuelve el empleado con mayor salario.

- **`__len__`** — número de empleados.
- **`__contains__`** — permite `empleado in departamento`.
- **`__str__`** — `"Departamento Desarrollo (3 empleados, media: 2.766,67 €)"`.

class Departamento:
    def __init__(self, nombre):
        self.nombre     = nombre
        self._empleados = []

    def añadir(self, empleado):
        if not isinstance(empleado, Empleado):
            raise TypeError(f"Se esperaba un Empleado, no {type(empleado).__name__}")
        if empleado in self._empleados:
            print(f"Aviso: {empleado.nombre_completo()} ya está en el departamento")
            return
        self._empleados.append(empleado)

    def eliminar(self, id_empleado):
        empleado = self.buscar(id_empleado)
        if empleado is None:
            raise ValueError(f"No existe ningún empleado con id '{id_empleado}'")
        self._empleados.remove(empleado)
        return empleado

    def buscar(self, id_empleado):
        for emp in self._empleados:
            if emp.id == id_empleado:
                return emp
        return None

    def salario_medio(self):
        if not self._empleados:
            return 0.0
        return sum(e.salario for e in self._empleados) / len(self._empleados)

    def el_mas_pagado(self):
        if not self._empleados:
            return None
        return max(self._empleados)   # usa __lt__

    def listar(self):
        if not self._empleados:
            print(f"El departamento {self.nombre} no tiene empleados.")
            return
        print(f"\n{'─'*55}")
        print(f"  {self.nombre.upper()} — {len(self._empleados)} empleado(s)")
        print(f"{'─'*55}")
        for emp in sorted(self._empleados, key=lambda e: e.apellido):
            print(f"  {emp}")
        print(f"{'─'*55}")
        print(f"  Salario medio: {self.salario_medio():,.2f} €")
        mas_pagado = self.el_mas_pagado()
        print(f"  Más pagado:    {mas_pagado.nombre_completo()} ({mas_pagado.salario:,.2f} €)")

    # ─── Dunder ───────────────────────────────────────────
    def __len__(self):
        return len(self._empleados)

    def __contains__(self, empleado):
        return empleado in self._empleados

    def __str__(self):
        media = self.salario_medio()
        return (
            f"Departamento {self.nombre} "
            f"({len(self)} empleado(s), media: {media:,.2f} €)"
        )


# --- Prueba ---
Empleado._contador_ids = 0

dev = Departamento("Desarrollo")
mkt = Departamento("Marketing")

e1 = Empleado("Ana",   "García",  "Desarrollo", 2500)
e2 = Empleado("Luis",  "Pérez",   "Marketing",  3100)
e3 = Empleado("Marta", "López",   "Desarrollo", 2800)
e4 = Empleado("Eva",   "Martín",  "Desarrollo", 2700)

dev.añadir(e1); dev.añadir(e3); dev.añadir(e4)
mkt.añadir(e2)

dev.listar()

print(f"\n{dev}")
print(f"¿e1 en dev? {e1 in dev}")
print(f"¿e2 en dev? {e2 in dev}")

dev.eliminar(e3.id)
print(f"\nTras eliminar {e3.nombre_completo()}: {dev}")
```

**Puntos clave:**
- `eliminar` usa `buscar` internamente — evita duplicar la lógica de búsqueda.
- `max(self._empleados)` funciona gracias a `__lt__` definido en `Empleado`.
- `sorted(..., key=lambda e: e.apellido)` ordena alfabéticamente sin modificar la lista interna.
- `__contains__` delega en `list.__contains__`, que a su vez usa `__eq__` de `Empleado` — la cadena de dunder funciona sola.

import json
from pathlib import Path

class Empresa:
    def __init__(self, nombre):
        self.nombre        = nombre
        self._departamentos = {}   # {nombre_dept: Departamento}

    # ─── Gestión de empleados ─────────────────────────────
    def contratar(self, empleado, nombre_departamento):
        if nombre_departamento not in self._departamentos:
            self._departamentos[nombre_departamento] = Departamento(nombre_departamento)
        self._departamentos[nombre_departamento].añadir(empleado)
        print(f"✅ {empleado.nombre_completo()} contratado/a en {nombre_departamento}")

    def transferir(self, id_empleado, nuevo_departamento):
        empleado, dept_origen = self.buscar_global(id_empleado)
        if empleado is None:
            raise ValueError(f"No existe ningún empleado con id '{id_empleado}'")
        dept_origen.eliminar(id_empleado)
        self.contratar(empleado, nuevo_departamento)
        print(f"🔄 {empleado.nombre_completo()} transferido/a de {dept_origen.nombre} a {nuevo_departamento}")

    def despedir(self, id_empleado):
        empleado, dept = self.buscar_global(id_empleado)
        if empleado is None:
            raise ValueError(f"No existe ningún empleado con id '{id_empleado}'")
        dept.eliminar(id_empleado)
        print(f"❌ {empleado.nombre_completo()} dado/a de baja en {dept.nombre}")
        return empleado

    def buscar_global(self, id_empleado):
        for dept in self._departamentos.values():
            emp = dept.buscar(id_empleado)
            if emp is not None:
                return emp, dept
        return None, None

    # ─── Informes ─────────────────────────────────────────
    def todos_los_empleados(self):
        return [emp for dept in self._departamentos.values() for emp in dept._empleados]

    def resumen(self):
        todos   = self.todos_los_empleados()
        n_total = len(todos)
        print(f"\n{'═'*60}")
        print(f"  {self.nombre.upper()}")
        print(f"  {n_total} empleado(s) en {len(self._departamentos)} departamento(s)")
        print(f"{'═'*60}")

        for dept in self._departamentos.values():
            dept.listar()

        if todos:
            mas_pagado = max(todos)   # usa __lt__
            media      = sum(e.salario for e in todos) / n_total
            print(f"\n  ► Salario medio empresa: {media:,.2f} €")
            print(f"  ► Más pagado/a:          {mas_pagado.nombre_completo()} ({mas_pagado.salario:,.2f} €)")
        print(f"{'═'*60}")

    def exportar_json(self, ruta):
        datos = {
            "empresa": self.nombre,
            "departamentos": {
                nombre: [emp.info() for emp in dept._empleados]
                for nombre, dept in self._departamentos.items()
            }
        }
        Path(ruta).write_text(
            json.dumps(datos, indent=2, ensure_ascii=False),
            encoding="utf-8"
        )
        print(f"\n📁 Datos exportados a {ruta}")

    def __str__(self):
        n = len(self.todos_los_empleados())
        return f"{self.nombre} ({len(self._departamentos)} depts., {n} empleados)"


# ─── Escenario de prueba ──────────────────────────────────

empresa = Empresa("TechGalicia S.L.")

    Empleado("Jorge",   "Fernández", "Desarrollo", 3400),
]

asignaciones = [
    (empleados[0], "Desarrollo"),
    (empleados[1], "Marketing"),
    (empleados[2], "Desarrollo"),
    (empleados[3], "Ventas"),
    (empleados[4], "Marketing"),
    (empleados[5], "Desarrollo"),
]

print("── Contrataciones ──")
for emp, dept in asignaciones:
    empresa.contratar(emp, dept)

print("\n── Transferencia ──")
empresa.transferir(empleados[4].id, "Desarrollo")

print("\n── Despido ──")
empresa.despedir(empleados[3].id)

empresa.resumen()
empresa.exportar_json("empresa.json")
```

**Puntos clave:**

### Anatomía de una clase

```python
class MiClase:
    atributo_clase = valor               # compartido

    def __init__(self, param):           # constructor
        self.atributo = param            # de instancia
        self._privado = None             # convención: privado

    @property
    def privado(self):                   # getter
        return self._privado

    @privado.setter
    def privado(self, valor):            # setter con validación
        self._privado = valor

    def metodo(self):                    # de instancia
        return self.atributo

    @classmethod
    def desde_dict(cls, datos):          # constructor alternativo
        return cls(datos["param"])

    @staticmethod
    def es_valido(valor):                # utilidad sin self ni cls
        return valor > 0

    def __str__(self):                   # para print() y str()
        return f"MiClase({self.atributo})"

    def __eq__(self, otro):              # para ==
        return self.atributo == otro.atributo
```

### Tabla de referencia rápida

| Concepto | Sintaxis | Uso |
|----------|---------|-----|
| Clase | `class Nombre:` | Plantilla |
| Instancia | `obj = Nombre(args)` | Copia concreta |
| Constructor | `def __init__(self, ...)` | Inicializar atributos |
| Atributo instancia | `self.x = valor` | Datos del objeto |
| Atributo clase | `x = valor` (fuera de métodos) | Datos compartidos |
| Método instancia | `def f(self)` | Trabaja con el objeto |
| Método de clase | `@classmethod def f(cls)` | Trabaja con la clase |
| Método estático | `@staticmethod def f()` | Utilidad sin estado |
| Propiedad | `@property def x(self)` | Getter con lógica |
| Setter | `@x.setter def x(self, v)` | Validar asignación |
