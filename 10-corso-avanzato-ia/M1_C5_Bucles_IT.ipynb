## 1. Perché abbiamo bisogno dei cicli?

Immagina di voler stampare i numeri da 1 a 5. Senza cicli:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

Funziona. Ma se fossero da 1 a 1.000? O da 1 a 1.000.000?

Con un ciclo, uno qualsiasi di questi casi è esattamente lo stesso codice — cambia solo il numero finale.

Ci sono **due tipi di ciclo** in Python:

| Ciclo | Quando usarlo |
|-------|---------------|
| `while` | Quando non sai quante volte ripeterai — ripeti **finché** si verifica una condizione |
| `for` | Quando sai quante volte ripeterai — scorri una sequenza di valori |

## 2. `while` — Ripetere finché si verifica una condizione

### Sintassi

```python
while condizione:
    # blocco che si ripete finché condizione è True
```

Python valuta la condizione **prima di ogni iterazione**. Se è `True`, esegue il blocco e la ricontrolla. Se è `False`, esce dal ciclo.

```
┌─────────────────────────────┐
│   condizione è True?        │
│        │                    │
│    Sì  │        No          │
│        ▼          ▼         │
│   esegue       esce dal     │
│   blocco       ciclo        │
│       │                     │
│       └──────── ritorna ────┘
```

> ⚠️ **Ciclo infinito:** se la condizione non diventa mai `False`, il ciclo non termina. Assicurati sempre che qualcosa all'interno del ciclo modifichi la condizione.

````python
# Contare da 1 a 5
contatore = 1

while contatore <= 5:
    print(contatore)
    contatore += 1     # senza questa riga → ciclo infinito

print("Ciclo terminato")
````

````python
# Chiedere un numero positivo finché l'utente non lo inserisce correttamente
numero = -1   # valore iniziale che forza l'ingresso nel ciclo

while numero <= 0:
    numero = int(input("Inserisci un numero positivo: "))
    if numero <= 0:
        print("Quel numero non è positivo, riprova.")

print(f"Perfetto, hai inserito: {numero}")
````

````python
# Conto alla rovescia
conto = 5

while conto > 0:
    print(f"{conto}...")
    conto -= 1

print("Decollo! 🚀")
````

### `while` con `else`

Python permette di aggiungere un blocco `else` a un `while`. Viene eseguito **quando il ciclo termina in modo naturale** (la condizione diventa `False`), ma **non** se si esce con `break`.

````python
tentativi = 3

while tentativi > 0:
    chiave = input("Inserisci la password: ")
    if chiave == "python":
        print("Accesso consentito ✅")
        break               # esce dal ciclo senza passare per l'else
    tentativi -= 1
    print(f"Password errata. Tentativi rimasti: {tentativi}")
else:
    # ci arriva solo se il while è terminato senza break (tentativi esauriti)
    print("Account bloccato ❌")
````

## 3. `for` — Scorrere una sequenza

Il ciclo `for` scorre uno a uno gli elementi di una **sequenza** (una lista, un range, una stringa…) ed esegue il blocco per ogni elemento.

### Sintassi

```python
for variabile in sequenza:
    # blocco eseguito una volta per ogni elemento
    # 'variabile' assume il valore dell'elemento corrente
```

A ogni iterazione, `variabile` **assume automaticamente** il valore dell'elemento successivo della sequenza.

```
sequenza = [10, 20, 30]

iterazione 1: variabile = 10  → esegue blocco
iterazione 2: variabile = 20  → esegue blocco
iterazione 3: variabile = 30  → esegue blocco
             (fine sequenza → esce)
```

````python
# Scorrere una lista di elementi
frutti = ["mela", "arancia", "banana", "uva"]

for frutto in frutti:
    print(f"Frutto: {frutto}")
````

````python
# Anche una stringa è una sequenza: viene scorsa lettera per lettera
parola = "Python"

for lettera in parola:
    print(lettera)
````

### 3.1 `range()` — Generare sequenze di numeri

`range()` genera una sequenza di numeri senza occupare memoria. È il modo più comune per controllare quante volte si ripete un `for`.

| Forma | Genera | Esempio |
|-------|--------|---------|
| `range(n)` | 0, 1, 2, … n-1 | `range(5)` → 0 1 2 3 4 |
| `range(inizio, fine)` | inizio, …, fine-1 | `range(2, 6)` → 2 3 4 5 |
| `range(inizio, fine, passo)` | da inizio a fine con salti | `range(0, 10, 2)` → 0 2 4 6 8 |

> Il valore `fine` **non viene mai incluso** — proprio come gli indici delle stringhe.

````python
# range(n): da 0 a n-1
print("range(5):")
for i in range(5):
    print(i, end=" ")   # end=" " stampa sulla stessa riga
print()

# range(inizio, fine)
print("range(1, 6):")
for i in range(1, 6):
    print(i, end=" ")
print()

# range(inizio, fine, passo)
print("range(0, 20, 3):")
for i in range(0, 20, 3):
    print(i, end=" ")
print()

# Passo negativo: conta all'indietro
print("range(10, 0, -2):")
for i in range(10, 0, -2):
    print(i, end=" ")
print()
````

````python
# Tabellina del 7
numero = 7

print(f"Tabellina del {numero}:")
for i in range(1, 11):
    print(f"  {numero} × {i:2} = {numero * i:3}")
````

### 3.2 `enumerate()` — Indice e valore insieme

Quando ti servono sia la **posizione** che il **valore** di ogni elemento, usa `enumerate()`. Evita il pattern `for i in range(len(lista))`, che è meno leggibile.

````python
materie = ["Matematica", "Italiano", "Storia", "Fisica"]

# Senza enumerate (meno idiomatico)
for i in range(len(materie)):
    print(f"{i + 1}. {materie[i]}")

print("---")

# Con enumerate (più chiaro)
for posizione, materia in enumerate(materie, start=1):
    print(f"{posizione}. {materia}")
````

### 3.3 Cicli `for` annidati

Un `for` dentro un altro `for`. Per ogni iterazione del ciclo esterno, il ciclo interno completa **tutte** le sue iterazioni.

````python
# Tabellina completa (da 1 a 5)
for i in range(1, 6):           # ciclo esterno: righe
    for j in range(1, 6):       # ciclo interno: colonne
        print(f"{i*j:3}", end=" ")
    print()                     # a capo al termine di ogni riga
````

## 4. `break` — Uscire dal ciclo prima del previsto

`break` **interrompe immediatamente il ciclo** e salta alla prima istruzione dopo il ciclo, ignorando le iterazioni rimanenti.

```
for / while ...:
    ...
    if condizione_di_uscita:
        break           ← esce dal ciclo
    ...
codice dopo il ciclo   ← continua qui
```

Si usa quando troviamo ciò che cercavamo e non ha più senso continuare a iterare.

````python
# Cercare il primo numero divisibile per 7 tra 1 e 100
for numero in range(1, 101):
    if numero % 7 == 0:
        print(f"Primo multiplo di 7: {numero}")
        break   # l'abbiamo già trovato, non serve continuare
    print(f"{numero} non è multiplo di 7")
````

````python
# Cercare una lettera in un testo
testo  = "programmazione"
cerca  = "a"
trovato = False

for i, lettera in enumerate(testo):
    if lettera == cerca:
        print(f"'{cerca}' trovata alla posizione {i}")
        trovato = True
        break

if not trovato:
    print(f"'{cerca}' non è presente nel testo")
````

## 5. `continue` — Saltare al ciclo successivo

`continue` **salta il resto del blocco corrente** e passa direttamente all'iterazione successiva del ciclo. Non esce dal ciclo — salta solo ciò che resta di quell'iterazione.

```
for / while ...:
    ...
    if condizione_di_salto:
        continue    ← salta al ciclo successivo
    ...             ← questo NON viene eseguito se continue è stato attivato
```

Si usa per **filtrare** elementi che non vogliamo elaborare.

````python
# Stampare solo i numeri dispari da 1 a 10
for n in range(1, 11):
    if n % 2 == 0:
        continue    # salta i pari
    print(n, end=" ")
print()
````

````python
# Elaborare una lista di voti, ignorando quelli non validi
voti = [8.5, -1, 6.0, 11, 7.3, 4.5, 9.1, -3, 5.0]

voti_validi = 0
somma       = 0

for voto in voti:
    if voto < 0 or voto > 10:
        print(f"  Voto non valido ignorato: {voto}")
        continue    # salta questo voto
    voti_validi += 1
    somma       += voto

if voti_validi > 0:
    media = somma / voti_validi
    print(f"\nVoti validi: {voti_validi}")
    print(f"Media: {media:.2f}")
````

### Differenza visiva tra `break` e `continue`

```
Lista: [1, 2, 3, 4, 5]
Condizione: numero == 3

Con BREAK:
  1 ✓  2 ✓  3 → SI FERMA  (4 e 5 non vengono elaborati)

Con CONTINUE:
  1 ✓  2 ✓  3 → SALTA  4 ✓  5 ✓  (3 non viene elaborato, ma 4 e 5 sì)
```

````python
numeri = [1, 2, 3, 4, 5]

print("Con break:")
for n in numeri:
    if n == 3:
        break
    print(n, end=" ")
print()

print("Con continue:")
for n in numeri:
    if n == 3:
        continue
    print(n, end=" ")
print()
````

## 6. Pattern comuni con i cicli

Ci sono tre pattern che ricorrono costantemente — vale la pena riconoscerli:

````python
# PATTERN 1: CONTATORE
# Conta quanti elementi soddisfano una condizione

voti = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
promossi = 0

for voto in voti:
    if voto >= 5:
        promossi += 1

print(f"Promossi: {promossi} su {len(voti)}")
````

````python
# PATTERN 2: ACCUMULATORE
# Accumula un valore (somma, prodotto, costruzione di stringa…)

voti  = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
somma = 0

for voto in voti:
    somma += voto

media = somma / len(voti)
print(f"Somma: {somma}")
print(f"Media: {media:.1f}")

# Python ha funzioni integrate per questo:
print(f"Con sum(): {sum(voti)}")
print(f"Con max(): {max(voti)}")
print(f"Con min(): {min(voti)}")
````

````python
# PATTERN 3: RICERCA
# Trova il primo che soddisfa una condizione

voti = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
primo_voto_alto = None

for voto in voti:
    if voto >= 9:
        primo_voto_alto = voto
        break

if primo_voto_alto is not None:
    print(f"Primo voto >= 9: {primo_voto_alto}")
else:
    print("Nessun voto >= 9")
````

### Quando usare `while` e quando `for`?

| Situazione | Usa |
|-----------|-----|
| Scorrere una lista, stringa o range | `for` |
| Ripetere esattamente N volte | `for` con `range(N)` |
| Ripetere finché l'utente non inserisce un valore valido | `while` |
| Ripetere finché non accade qualcosa (senza sapere quando) | `while` |
| Ciclo infinito controllato con `break` | `while True:` |

Il pattern `while True:` con `break` è molto comune per **menu** o **cicli di gioco**:

````python
# Pattern while True: il ciclo esce solo quando l'utente sceglie di uscire
while True:
    opzione = input("\nMenu: (1) Saluta  (2) Congeda  (3) Esci → ")

    if opzione == "1":
        print("Ciao!")
    elif opzione == "2":
        print("Arrivederci!")
    elif opzione == "3":
        print("Uscita in corso...")
        break
    else:
        print("Opzione non valida")
````

### Esercizio 1 — Indovinello base

**Consegna:** Scrivi un programma che:
1. Abbia un numero segreto fisso (per esempio, 42).
2. Chieda all'utente di indovinarlo.
3. Ripeta finché non indovina, dando indizi: *"troppo alto"* o *"troppo basso"*.
4. Quando indovina, mostri quanti tentativi ha impiegato.

**Esempio di output atteso:**
```
Indovina il numero (tra 1 e 100): 50
Troppo alto.
Indovina il numero (tra 1 e 100): 25
Troppo basso.
Indovina il numero (tra 1 e 100): 42
Corretto! Lo hai indovinato in 3 tentativi.
```

````python
# ✏️ IL TUO CODICE QUI
# Il numero segreto è 42
````

<details>
<summary>💡 Vedi soluzione — Esercizio 1</summary>

```python
segreto   = 42
tentativi = 0

while True:
    tentativo  = int(input("Indovina il numero (tra 1 e 100): "))
    tentativi += 1

    if tentativo > segreto:
        print("Troppo alto.")
    elif tentativo < segreto:
        print("Troppo basso.")
    else:
        print(f"Corretto! Lo hai indovinato in {tentativi} tentativi.")
        break
```

**Punti chiave:**
- `while True:` crea un ciclo infinito che esce solo quando l'utente indovina.
- Incrementiamo `tentativi` all'inizio di ogni giro, prima di controllare la risposta.
- Il `break` viene raggiunto solo quando `tentativo == segreto` — l'`else` dell'`if/elif/else` garantisce che sia il caso di successo.
</details>

**Esempio di output (sconfitta):**
```
Tentativo 1/6 → 50: Troppo alto.
Tentativo 2/6 → 20: Troppo basso.
Tentativo 3/6 → 35: Troppo alto.
Tentativo 4/6 → 28: Troppo basso.
```

```python
segreto        = 30
max_tentativi  = 6
vinto          = False

for numero_tentativo in range(1, max_tentativi + 1):
```

- Alternativa: usare `for/else` — il blocco `else` del `for` viene eseguito se il ciclo è terminato senza `break`.

### Esercizio 4 — Cronologia dei tentativi

**Consegna:** Aggiungi una **cronologia** dei tentativi: salva ogni numero inserito (valido) in una variabile di tipo stringa e mostrala alla fine, indicando se ciascuno era alto, basso o corretto.

**Esempio di output finale:**
```
--- Cronologia ---
  Tentativo 1: 50  → alto
  Tentativo 2: 25  → basso
  Tentativo 3: 37  → corretto!
```

> 💡 Salva la cronologia come una variabile di testo (`str`) che costruisci con `+=` a ogni iterazione.

````python
# ✏️ IL TUO CODICE QUI
import random
````

<details>
<summary>💡 Vedi soluzione — Esercizio 4</summary>

```python
import random

segreto       = random.randint(1, 100)
max_tentativi = 6
tentativi     = 0
vinto         = False
cronologia    = ""   # accumulatore di testo

print("Ho pensato un numero tra 1 e 100.")

while tentativi < max_tentativi and not vinto:
    tentativo = int(input(f"Tentativo {tentativi + 1}/{max_tentativi} → "))

    if tentativo < 1 or tentativo > 100:
        print("⚠️ Fuori range.")
        continue

    tentativi += 1

    if tentativo > segreto:
        risultato = "alto"
        print("Troppo alto.")
    elif tentativo < segreto:
        risultato = "basso"
        print("Troppo basso.")
    else:
        risultato = "corretto!"
        vinto     = True

    cronologia += f"  Tentativo {tentativi}: {tentativo:3}  → {risultato}\n"

# Risultato finale
if vinto:
    print(f"\nCe l'hai fatta in {tentativi} tentativi!")
else:
    print(f"\nTentativi esauriti! Il numero era {segreto}.")

print("\n--- Cronologia ---")
print(cronologia)
```

**Punti chiave:**
- `cronologia` è una stringa vuota che costruiamo con `+=` — pattern accumulatore applicato al testo.
- `\n` alla fine di ogni riga della cronologia produce l'a capo corretto quando la si stampa.
- `{tentativo:3}` allinea i numeri a destra in un campo di 3 caratteri — rende la cronologia più leggibile.
</details>

### Esercizio 5 (sfida) — Gioco completo con più partite

**Consegna:** Aggiungi un ciclo esterno che permetta di **giocare più partite**. Al termine di ogni partita, chiedi se il giocatore vuole rigiocare. All'uscita, mostra le statistiche globali:

- Partite giocate
- Partite vinte
- Partite perse
- Percentuale di vittorie
- Miglior risultato (meno tentativi in una vittoria)

**Esempio di output finale:**
```
Vuoi giocare un'altra partita? (s/n): n

=== Statistiche finali ===
Partite giocate:    3
Vittorie:           2
Sconfitte:          1
% di vittorie:      66.67%
Miglior risultato:  3 tentativi
```

````python
# ✏️ IL TUO CODICE QUI
import random
````

<details>
<summary>💡 Vedi soluzione — Esercizio 5</summary>

```python
import random

# Statistiche globali
partite_giocate = 0
vittorie        = 0
miglior_risultato = None   # None finché non c'è almeno una vittoria

while True:   # ciclo delle partite
    segreto       = random.randint(1, 100)
    max_tentativi = 6
    tentativi     = 0
    vinto         = False

    print(f"\n=== Partita {partite_giocate + 1} ===")
    print("Ho pensato un numero tra 1 e 100.")

    while tentativi < max_tentativi and not vinto:
        tentativo = int(input(f"Tentativo {tentativi + 1}/{max_tentativi} → "))

        if tentativo < 1 or tentativo > 100:
            print("⚠️ Fuori range.")
            continue

        tentativi += 1

        if tentativo > segreto:
            print("Troppo alto.")
        elif tentativo < segreto:
            print("Troppo basso.")
        else:
            vinto = True

    # Risultato della partita
    partite_giocate += 1

    if vinto:
        vittorie += 1
        print(f"Corretto! Ce l'hai fatta in {tentativi} tentativi.")
        # Aggiorna il miglior risultato
        if miglior_risultato is None or tentativi < miglior_risultato:
            miglior_risultato = tentativi
    else:
        print(f"Tentativi esauriti! Il numero era {segreto}.")

    # Un'altra partita?
    altra = input("\nVuoi giocare un'altra partita? (s/n): ")
    if altra.lower() != "s":
        break

# Statistiche finali
sconfitte  = partite_giocate - vittorie
percentuale = (vittorie / partite_giocate) * 100 if partite_giocate > 0 else 0

print("\n=== Statistiche finali ===")
print(f"Partite giocate:    {partite_giocate}")
print(f"Vittorie:           {vittorie}")
print(f"Sconfitte:          {sconfitte}")
print(f"% di vittorie:      {percentuale:.2f}%")

if miglior_risultato is not None:
    print(f"Miglior risultato:  {miglior_risultato} tentativi")
else:
    print("Miglior risultato:  nessuna vittoria")
```

**Punti chiave:**
- Due cicli annidati: quello esterno controlla le partite, quello interno controlla ogni tentativo.
- `miglior_risultato = None` all'inizio permette di distinguere tra "ancora nessuna vittoria" e "vittoria con 0 tentativi" (impossibile, ma è buona pratica).
- L'aggiornamento del miglior risultato usa `is None or tentativi < miglior_risultato` — la prima condizione copre la prima vittoria.
- `percentuale` usa l'operatore ternario per evitare la divisione per zero se `partite_giocate == 0`.
