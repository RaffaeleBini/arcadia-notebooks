## 2. Estructura del sistema de archivos Linux

En Linux (y macOS), todo cuelga de un único punto: la **raíz `/`**.

```
/                          ← raíz (root)
├── bin/                   ← ejecutables esenciales del sistema (ls, cp, mv…)
├── etc/                   ← archivos de configuración del sistema
├── home/                  ← directorios personales de los usuarios
│   ├── ana/               ← directorio personal de ana  (~ para ana)
│   └── carlos/            ← directorio personal de carlos
├── tmp/                   ← archivos temporales (se borran al reiniciar)
├── usr/                   ← programas instalados por el usuario
│   └── local/
│       └── bin/           ← programas instalados manualmente
├── var/                   ← datos variables: logs, bases de datos…
│   └── log/               ← logs del sistema
└── dev/                   ← dispositivos (disco, USB, terminal…)
```

**En macOS** la estructura es similar pero con diferencias:
- Los directorios personales están en `/Users/`, no en `/home/`.
- Los programas instalados con Homebrew van a `/opt/homebrew/` (Apple Silicon) o `/usr/local/` (Intel).

**`~` (tilde)** es un atajo para tu directorio personal: `~/proyectos` equivale a `/home/tu-usuario/proyectos`.

## 3. Rutas absolutas y relativas

Una **ruta** es la dirección de un archivo o directorio en el sistema.

```
RUTA ABSOLUTA                      RUTA RELATIVA
─────────────────────────────      ────────────────────────────────
Empieza siempre con /              Empieza desde donde estás ahora
/home/ana/proyectos/curso          proyectos/curso
Funciona desde cualquier lugar     Depende del directorio actual
Más larga de escribir              Más corta
```

**Atajos de ruta:**

| Atajo | Significa |
|-------|-----------|
| `.`   | El directorio actual |
| `..`  | El directorio padre (un nivel arriba) |
| `~`   | Tu directorio personal (`/home/usuario`) |
| `-`   | El directorio anterior (donde estabas antes) |

```bash
# Ejemplos con los atajos:
cd ..          # subir un nivel
cd ../..       # subir dos niveles
cd ~/proyectos # ir a proyectos en tu home, desde cualquier lugar
cd -           # volver al directorio anterior
```

## 5. `ls` — ¿Qué hay aquí?

`ls` (*list*) muestra el contenido de un directorio.

```bash
ls                    # contenido del directorio actual
ls /ruta/directorio   # contenido de otro directorio
ls -l                 # formato largo (permisos, tamaño, fecha)
ls -a                 # incluir archivos ocultos (empiezan por .)
ls -la                # largo + ocultos (las flags se combinan)
ls -lh                # largo + tamaños legibles (KB, MB, GB)
ls -lt                # ordenar por fecha de modificación
ls -lS                # ordenar por tamaño
ls *.py               # solo archivos .py (glob)
```

````python
import os, shutil
from pathlib import Path

# Crear un directorio de demo con estructura
DEMO = Path.home() / "linux_demo"
if DEMO.exists(): shutil.rmtree(DEMO)

# Crear estructura
(DEMO / "proyectos" / "curso_ia").mkdir(parents=True)
(DEMO / "proyectos" / "trabajo").mkdir(parents=True)
(DEMO / "datos").mkdir()
(DEMO / ".config").mkdir()  # carpeta oculta

(DEMO / "notas.txt").write_text("Mis notas\n", encoding="utf-8")
(DEMO / "script.py").write_text("print('hola')\n", encoding="utf-8")
(DEMO / "datos" / "dataset.csv").write_text("id,valor\n1,100\n", encoding="utf-8")
(DEMO / ".config" / "settings.json").write_text("{}", encoding="utf-8")

os.chdir(DEMO)
print(f"Directorio de demo: {DEMO}")
````

````python
# ls básico
print("=== ls (básico) ===")
!ls

print("\n=== ls -l (formato largo) ===")
!ls -l

print("\n=== ls -la (con archivos ocultos) ===")
!ls -la
````

````python
# Entender el formato largo de ls -l
print("""
drwxr-xr-x  3  ana  staff   96  ene 15 10:30  proyectos
│└─┬──┘└─┬┘ │  │    │       │   │              └─ nombre
│  │     │  │  │    │       │   └─ fecha de modificación
│  │     │  │  │    │       └─ tamaño en bytes
│  │     │  │  │    └─ grupo propietario
│  │     │  │  └─ usuario propietario
│  │     │  └─ número de enlaces
│  │     └─ permisos del grupo
│  └─ permisos del usuario propietario
└─ tipo: d=directorio, -=archivo, l=enlace simbólico
""")

print("\n=== ls -lh (tamaños legibles) ===")
!ls -lh datos/
````

## 6. `cd` — Navegar entre directorios

`cd` (*change directory*) te mueve a otro directorio.

```bash
cd /ruta/absoluta     # ir a una ruta absoluta
cd nombre_carpeta     # ir a una subcarpeta (relativa)
cd ..                 # subir un nivel
cd ../otra_carpeta    # subir y entrar en otra
cd ~                  # ir al home
cd                    # también va al home (sin argumentos)
cd -                  # volver al directorio anterior
```

````python
# Nota: en Jupyter, !cd no persiste entre celdas.
# Usa os.chdir() para mover el directorio de trabajo de Python,
# o usa %%bash para ejecutar varias líneas en el mismo shell.

print("Demostración de navegación:")
!pwd

print("\nEntrar en proyectos:")
os.chdir(DEMO / "proyectos")
!pwd

print("\nSubir un nivel (..):")
os.chdir(DEMO)
!pwd

print("\nRuta relativa compuesta (proyectos/curso_ia):")
os.chdir(DEMO / "proyectos" / "curso_ia")
!pwd

print("\nVolver al demo con ~ (home relativo):")
os.chdir(DEMO)
!pwd
````

````python
%%bash
# Con %%bash, todas las líneas se ejecutan en el mismo shell
# así cd sí persiste dentro de este bloque

cd ~/linux_demo
echo "1. Estamos en: $(pwd)"

cd proyectos
echo "2. Entramos en: $(pwd)"

cd curso_ia
echo "3. Entramos en: $(pwd)"

cd ../..
echo "4. Subimos dos niveles: $(pwd)"

cd -
echo "5. Volvemos al anterior: $(pwd)"
````

## 7. `mkdir` y `rmdir` — Crear y borrar directorios


**`-p`** (*parents*) es el flag más útil: crea todos los directorios intermedios que no existan.

## 8. `tree` — Ver la estructura completa

`tree` muestra el árbol de directorios de forma visual. Muy útil para entender la estructura de un proyecto.

```bash
tree                  # árbol completo desde el directorio actual
tree -L 2             # solo hasta 2 niveles de profundidad
tree -a               # incluir archivos ocultos
tree -d               # solo directorios (sin archivos)
tree --noreport       # sin la línea de resumen al final
```

> Si `tree` no está instalado: `brew install tree` (macOS) o `apt install tree` (Linux).

````python
# tree (si está instalado)
!tree -L 2 2>/dev/null || echo "tree no instalado — usando alternativa Python"

# Alternativa Python cuando tree no está disponible
def mostrar_arbol(ruta, prefijo="", max_nivel=3, nivel=0):
    if nivel > max_nivel:
        return
    ruta = Path(ruta)
    contenido = sorted(ruta.iterdir(), key=lambda p: (p.is_file(), p.name))
    for i, item in enumerate(contenido):
        es_ultimo = i == len(contenido) - 1
        conector = "└── " if es_ultimo else "├── "
        tipo = "/" if item.is_dir() else ""
        print(f"{prefijo}{conector}{item.name}{tipo}")
        if item.is_dir():
            extension = "    " if es_ultimo else "│   "
            mostrar_arbol(item, prefijo + extension, max_nivel, nivel + 1)

print(f"\n{DEMO.name}/")
mostrar_arbol(DEMO)
````

**Enunciado:** Escribe la función Python `explorar_proyecto(ruta)` que genere un informe completo de un directorio:

- Los cinco archivos más grandes con su tamaño legible (KB/MB).
- Archivos ocultos encontrados.
- Profundidad máxima del árbol.

<details>
<summary>💡 Ver solución — Ejercicio 5</summary>

```python
from pathlib import Path
from collections import Counter

def formato_bytes(n):
    for unidad in ("B", "KB", "MB", "GB"):
        if n < 1024:
            return f"{n:.1f} {unidad}"
        n /= 1024
    return f"{n:.1f} TB"

def explorar_proyecto(ruta):
    ruta = Path(ruta).expanduser().resolve()

    archivos = [p for p in ruta.rglob("*") if p.is_file()]
    directorios = [p for p in ruta.rglob("*") if p.is_dir()]
    ocultos = [p for p in archivos if p.name.startswith(".")]

    extensiones = Counter(p.suffix.lower() or "(sin extensión)" for p in archivos)

    mas_grandes = sorted(archivos, key=lambda p: p.stat().st_size, reverse=True)[:5]

    def profundidad(p):
        return len(p.relative_to(ruta).parts)

    prof_max = max((profundidad(p) for p in archivos), default=0)

    print(f"\n{'═'*55}")
    print(f"  INFORME: {ruta.name}")
    print(f"{'═'*55}")
    print(f"  Archivos:     {len(archivos)}")
    print(f"  Directorios:  {len(directorios)}")
    print(f"  Profundidad:  {prof_max} niveles")

    print(f"\n  EXTENSIONES:")
    for ext, count in extensiones.most_common():
        print(f"    {ext:<20} {count:>3} archivo(s)")

    print(f"\n  5 ARCHIVOS MÁS GRANDES:")
    for p in mas_grandes:
        tam = formato_bytes(p.stat().st_size)
        rel = p.relative_to(ruta)
        print(f"    {tam:>10}  {rel}")

    print(f"\n  ARCHIVOS OCULTOS ({len(ocultos)}):")
    for p in ocultos:
        print(f"    {p.relative_to(ruta)}")
    if not ocultos:
        print("    Ninguno")

    print(f"{'═'*55}")


explorar_proyecto("~/ml_proyecto")
```
</details>
