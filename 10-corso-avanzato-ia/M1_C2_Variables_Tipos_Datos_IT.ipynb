## Perché sono importanti i tipi di dati?

Immagina di avere due variabili:

```python
a = "3"
b = "5"
print(a + b)   # → "35"  ← concatena testo
```

contro

```python
a = 3
b = 5
print(a + b)   # → 8  ← somma numeri
```

L'operatore `+` fa cose completamente diverse a seconda del **tipo** dei dati. Capire i tipi è essenziale per evitare errori difficili da individuare.

````python
# Esempio: lo stesso operatore, risultati molto diversi
a_testo   = "3"
b_testo   = "5"
a_numero = 3
b_numero = 5

print("Testo + testo:   ", a_testo  + b_testo)   # concatena
print("Numero + numero: ", a_numero + b_numero)   # somma
````

## 1. `int` — Numeri interi

Un `int` è qualsiasi numero senza parte decimale: positivo, negativo o zero.

```python
età          = 28
temperatura = -5
zero         = 0
popolazione  = 8_000_000_000   # puoi usare _ come separatore delle migliaia
```

### Fino a che numero può arrivare un `int`?

In Python 3, gli interi **non hanno limiti di dimensione**. Puoi rappresentare numeri astronomicamente grandi senza alcun problema. Questo non vale per tutti i linguaggi.

### Operazioni specifiche degli interi

| Operatore | Cosa fa | Esempio | Risultato |
|----------|----------|---------|----------|
| `//` | Divisione intera (senza decimali) | `17 // 5` | `3` |
| `%` | Modulo (resto della divisione) | `17 % 5` | `2` |
| `**` | Potenza | `2 ** 10` | `1024` |

````python
# Esempi con int
età          = 28
anno_corrente = 2025
popolazione  = 8_000_000_000   # il trattino basso non influisce sul valore, solo sulla leggibilità

print(type(età))
print(f"Età: {età}")
print(f"Anno di nascita: {anno_corrente - età}")
print(f"Popolazione mondiale: {popolazione:,}")   # :, formatta con virgole come separatore delle migliaia
````

````python
# Operazioni esclusive degli interi
a = 17
b = 5

print(f"{a} // {b} = {a // b}")   # quante volte b entra in a
print(f"{a} %  {b} = {a % b}")    # cosa avanza
print(f"2 ** 10 = {2 ** 10}")      # 2 elevato a 10

# Curiosità: int di dimensione arbitraria
numero_gigante = 2 ** 100
print(f"2 elevato a 100 = {numero_gigante}")
````

## 2. `float` — Numeri decimali

Un `float` è un numero con parte decimale. Internamente, Python lo rappresenta in notazione a virgola mobile (da qui il nome).

```python
prezzo   = 19.99
pi       = 3.14159
negativo = -0.5
scienza  = 1.5e6    # notazione scientifica: 1.5 × 10⁶ = 1 500 000
```

### ⚠️ Il problema della precisione dei float

I `float` hanno un limite di precisione. Questo può causare risultati inaspettati:

```python
0.1 + 0.2   # → 0.30000000000000004  (non è 0.3!)
```

Questo non è un bug di Python — è una limitazione di come i computer rappresentano i decimali in binario. Per calcoli finanziari precisi, si usa la libreria `decimal`.

````python
# Esempi con float
prezzo  = 19.99
pi      = 3.14159
scienza = 1.5e6     # notazione scientifica

print(type(prezzo))
print(f"Prezzo:   {prezzo}")
print(f"Pi:       {pi}")
print(f"Scienza:  {scienza}")        # mostra 1500000.0
print(f"Scienza:  {scienza:,.0f}")   # formattato con separatore delle migliaia
````

````python
# Il problema di precisione dei float
risultato = 0.1 + 0.2
print(f"0.1 + 0.2 = {risultato}")             # mostra il problema
print(f"0.1 + 0.2 = {risultato:.2f}")         # arrotondato a 2 decimali

# Confronto: non confrontare mai i float con ==
print(0.1 + 0.2 == 0.3)                       # False ← trappola classica
print(round(0.1 + 0.2, 2) == 0.3)             # True ← modo corretto

# round() arrotonda a N decimali
print(round(3.14159, 2))                       # 3.14
````

### Funzioni matematiche utili con i numeri

````python
import math

x = -7.8

print(f"abs({x})           = {abs(x)}")           # valore assoluto
print(f"round({x})         = {round(x)}")         # arrotondamento all'intero più vicino
print(f"math.floor({x})    = {math.floor(x)}")    # arrotondamento verso il basso
print(f"math.ceil({x})     = {math.ceil(x)}")     # arrotondamento verso l'alto
print(f"math.sqrt(144)     = {math.sqrt(144)}")   # radice quadrata
print(f"math.log10(1000)   = {math.log10(1000)}") # logaritmo in base 10
````

## 3. `str` — Stringhe di testo

Uno `str` (string) è una sequenza di caratteri. Si scrive tra apici singoli `'...'` o doppi `"..."`. Sono equivalenti.

```python
nome      = "Ana"
messaggio = 'Ciao, mondo'
vuota     = ""            # stringa vuota
multiriga = """Questo testo
occupa diverse
righe"""                # tre apici per il testo multiriga
```

### Le stringhe sono sequenze

Ogni carattere di una stringa ha una **posizione** (indice), a partire da 0:

```
  testo = "Python"
            P  y  t  h  o  n
  indice:   0  1  2  3  4  5
  negativo:-6 -5 -4 -3 -2 -1
```

Puoi accedere a un carattere con `testo[indice]` e a una parte con `testo[inizio:fine]` (**slicing**).

````python
testo = "Python"

# Accesso per indice
print(f"Primo carattere:   {testo[0]}")    # P
print(f"Ultimo carattere:  {testo[-1]}")   # n
print(f"Terzo carattere:   {testo[2]}")    # t

# Slicing: testo[inizio:fine]  (fine non incluso)
print(f"Primi 3:          {testo[0:3]}")  # Pyt
print(f"Dal 3:            {testo[3:]}")   # hon
print(f"Ultimi 3:         {testo[-3:]}")  # hon
print(f"Lunghezza:        {len(testo)}")  # 6
````

### Metodi delle stringhe

Le stringhe hanno molti **metodi** — funzioni che appartengono alla stringa stessa e si chiamano con un punto.

| Metodo | Cosa fa | Esempio |
|--------|----------|--------|
| `.upper()` | Converte in maiuscolo | `"ciao".upper()` → `"CIAO"` |
| `.lower()` | Converte in minuscolo | `"CIAO".lower()` → `"ciao"` |
| `.strip()` | Rimuove gli spazi all'inizio e alla fine | `" ciao ".strip()` → `"ciao"` |
| `.replace(a, b)` | Sostituisce `a` con `b` | `"ciao mondo".replace("mondo", "Python")` |
| `.split(sep)` | Divide la stringa in una lista | `"a,b,c".split(",")` → `['a', 'b', 'c']` |
| `.startswith(x)` | Inizia con `x`? | `"Python".startswith("Py")` → `True` |
| `.count(x)` | Quante volte compare `x` | `"banana".count("a")` → `3` |
| `.find(x)` | Posizione di `x` (-1 se non esiste) | `"Python".find("th")` → `2` |

````python
frase = "  Ciao, Mondo!  "

print(frase.strip())           # rimuove gli spazi
print(frase.strip().upper())   # concateniamo metodi
print(frase.strip().lower())
print(frase.strip().replace("Mondo", "Python"))
print(frase.strip().split(", "))  # divide per ", "

# Contare e cercare
testo = "banana"
print(f"La lettera 'a' compare {testo.count('a')} volte")
print(f"La posizione di 'nan' è {testo.find('nan')}")
````

````python
# Operazioni con le stringhe
saluto = "Ciao"
nome   = "Ana"

# Concatenazione con +
messaggio = saluto + ", " + nome + "!"
print(messaggio)

# Ripetizione con *
separatore = "-" * 30
print(separatore)

# f-string (il modo più consigliato)
print(f"{saluto}, {nome}!")

# Verificare se un testo contiene qualcosa (operatore 'in')
print("Ana" in messaggio)    # True
print("Pedro" in messaggio)  # False
````

## 4. `bool` — Valori logici

Un `bool` può avere solo due valori: `True` (vero) o `False` (falso). Sono la base di tutta la logica nella programmazione.

> ⚠️ Importante: in Python `True` e `False` si scrivono con la prima lettera maiuscola.

### Da dove vengono i booleani?

Nella maggior parte dei casi non li scrivi direttamente — sono il risultato di un **confronto** o una **condizione**:

````python
# Operatori di confronto → restituiscono sempre bool
print(5 > 3)      # True
print(5 < 3)      # False
print(5 == 5)     # True  (== è uguaglianza; = è assegnazione)
print(5 != 3)     # True  (!= è diverso)
print(5 >= 5)     # True
print(5 <= 4)     # False

# Con variabili
età = 20
è_maggiorenne = età >= 18
print(f"È maggiorenne? {è_maggiorenne}")
````

### Operatori logici: `and`, `or`, `not`

Combinano condizioni booleane:

| Operatore | Significato | Risultato |
|----------|-------------|----------|
| `A and B` | Entrambe devono essere `True` | `True` solo se A e B sono `True` |
| `A or B` | Almeno una deve essere `True` | `False` solo se A e B sono `False` |
| `not A` | Inverte il valore | `not True` → `False` |

```
Tavola di verità di AND:
  True  and True   → True
  True  and False  → False
  False and True   → False
  False and False  → False

Tavola di verità di OR:
  True  or True   → True
  True  or False  → True
  False or True   → True
  False or False  → False
```

````python
età          = 20
ha_carta_identità = True

# and: entrambe le condizioni devono essere vere
può_votare = età >= 18 and ha_carta_identità
print(f"Può votare? {può_votare}")

# or: basta che una sia vera
è_studente = False
è_pensionato = True
ha_sconto = è_studente or è_pensionato
print(f"Ha sconto? {ha_sconto}")

# not: inverte
print(f"Non è studente? {not è_studente}")
````

### Valori "truthy" e "falsy"

In Python, qualsiasi valore può essere usato in un contesto booleano. Alcuni valori vengono considerati automaticamente `False`:

| Valore | È `False`? |
|-------|-------------|
| `0` (intero zero) | Sì |
| `0.0` (float zero) | Sì |
| `""` (stringa vuota) | Sì |
| `None` (assenza di valore) | Sì |
| Qualsiasi altro valore | No (è `True`) |

````python
# bool() converte qualsiasi valore in booleano
print(bool(0))       # False
print(bool(0.0))     # False
print(bool(""))      # False
print(bool(None))    # False

print("---")

print(bool(1))       # True
print(bool(-5))      # True  (qualsiasi numero diverso da zero)
print(bool("ciao"))  # True  (qualsiasi stringa non vuota)
print(bool(0.001))   # True
````

## 5. Casting — Convertire i tipi

Il **casting** (o conversione di tipo) consiste nel trasformare un valore da un tipo a un altro.

In Python si fa con le funzioni di conversione:

| Funzione | Converte in | Esempio |
|---------|------------|--------|
| `int(x)` | Intero | `int("42")` → `42` |
| `float(x)` | Decimale | `float("3.14")` → `3.14` |
| `str(x)` | Testo | `str(100)` → `"100"` |
| `bool(x)` | Booleano | `bool(0)` → `False` |

### Quando ti serve?

Il caso più frequente: `input()` **restituisce sempre una stringa**, anche se l'utente scrive un numero. Devi convertirlo prima di operarci.

```python
dato = input("Scrivi un numero: ")   # dato è "42" (str)
numero = int(dato)                    # ora è 42 (int)
```

````python
# int() — converte in intero
print(int("42"))       # da str a int
print(int(3.9))        # da float a int (tronca, NON arrotonda)
print(int(True))       # True → 1
print(int(False))      # False → 0

print("---")

# float() — converte in decimale
print(float("3.14"))   # da str a float
print(float(10))       # da int a float
print(float("1e3"))    # notazione scientifica in stringa

print("---")

# str() — converte in testo
print(str(42))         # da int a str
print(str(3.14))       # da float a str
print(str(True))       # da bool a str
````

````python
# ⚠️ Conversioni che falliscono
# Decommenta ogni riga per vedere l'errore

# int("ciao")       # ValueError: non si può convertire testo non numerico a int
# int("3.14")       # ValueError: non si può convertire una float-string direttamente a int
# float("abc")      # ValueError

# Soluzione per "3.14" → int: passare prima per float
risultato = int(float("3.14"))
print(f"int(float('3.14')) = {risultato}")
````

````python
# Python fa alcune conversioni automatiche (implicite)
# int + float → float  (Python sceglie il tipo più preciso)
risultato = 3 + 1.5
print(risultato)          # 4.5
print(type(risultato))    # float

# bool nelle operazioni numeriche
print(True + True)        # 2  (True = 1)
print(True + 5)           # 6
print(False * 100)        # 0  (False = 0)
````

### Riepilogo visivo delle conversioni

```
         int()
  str ──────────────▶ int
  float ────(tronca)─▶ int
  bool ─────(0 o 1)──▶ int

         float()
  str ──────────────▶ float
  int ──────────────▶ float

         str()
  int   ────────────▶ str
  float ────────────▶ str
  bool  ────────────▶ str  ("True" / "False")

         bool()
  0, 0.0, "", None  ──▶ False
  qualsiasi altro   ──▶ True
```

### Esercizio 1 — Convertitore di temperatura

**Consegna:** Scrivi un programma che chieda una temperatura in **gradi Celsius** e la converta in:
- **Fahrenheit:** `F = C × 9/5 + 32`
- **Kelvin:** `K = C + 273.15`

**Esempio di output atteso:**
```
Inserisci la temperatura in Celsius: 100

100.00 °C equivale a:
  212.00 °F (Fahrenheit)
  373.15 K  (Kelvin)
```

### Esercizio 4 — Convertitore di tempo

**Consegna:** Scrivi un programma che chieda un numero di **secondi** (intero) e lo scomponga in:
- Giorni
- Ore rimanenti
- Minuti rimanenti
- Secondi rimanenti

**Suggerimento:** usa la divisione intera `//` e il modulo `%`.

**Esempio di output atteso:**
```
Inserisci il numero di secondi: 90061

90061 secondi equivale a:
  1 giorno/i
  1 ora/e
  1 minuto/i
  1 secondo/i
```

````python
# ✏️ IL TUO CODICE QUI
# Suggerimento:
#   1 minuto  = 60 secondi
#   1 ora     = 3600 secondi
#   1 giorno  = 86400 secondi
#
# Strategia: procedere dall'unità più grande alla più piccola
#   giorni   = totale // 86400
#   resto    = totale % 86400   ← i secondi che avanzano dopo aver tolto i giorni
#   ore      = resto // 3600   ← e così via
````

<details>
<summary>💡 Vedi soluzione — Esercizio 4</summary>

```python
totale = int(input("Inserisci il numero di secondi: "))

giorni  = totale // 86400
resto   = totale  % 86400

ore     = resto  // 3600
resto   = resto   % 3600

minuti  = resto  // 60
secondi = resto  % 60

print(f"\n{totale} secondi equivale a:")
print(f"  {giorni} giorno/i")
print(f"  {ore} ora/e")
print(f"  {minuti} minuto/i")
print(f"  {secondi} secondo/i")
```

**Punti chiave:**
- La strategia è sempre la stessa: `//` per ottenere la quantità intera dell'unità grande, `%` per ottenere il resto che passiamo all'unità più piccola successiva.
- Riutilizziamo la variabile `resto` a ogni passaggio — è valido perché non ci serve più il valore precedente.
- `int()` è necessario perché `input()` restituisce una stringa e gli operatori `//` e `%` richiedono interi.
</details>

### Esercizio 5 (sfida) — Convertitore completo con validazione di tipo

**Consegna:** Crea un convertitore di valuta che:
1. Chieda un importo in **euro** (`float`).
2. Usi questi tassi di cambio:
   - `1 EUR = 1.08 USD`
   - `1 EUR = 0.86 GBP`
   - `1 EUR = 162.5 JPY`
3. Mostri i risultati con il formato di ogni valuta:
   - USD e GBP con 2 decimali
   - JPY come numero intero (senza decimali, perché lo yen non ha centesimi)
4. Mostri anche se l'importo in USD supera i 1 000 dollari usando un `bool`.

**Esempio di output atteso:**
```
Inserisci l'importo in euro: 950

--- Conversione di 950.00 EUR ---
  USD: 1026.00 $
  GBP: 817.00 £
  JPY: 154375 ¥

Supera i 1000 USD? True
```

````python
# ✏️ IL TUO CODICE QUI
````

<details>
<summary>💡 Vedi soluzione — Esercizio 5</summary>

```python
euro = float(input("Inserisci l'importo in euro: "))

# Tassi di cambio
TC_USD = 1.08
TC_GBP = 0.86
TC_JPY = 162.5

# Conversioni
usd = euro * TC_USD
gbp = euro * TC_GBP
jpy = int(euro * TC_JPY)   # lo yen è intero

# Supera i 1000 USD?
supera_mille = usd > 1000     # questo è un bool

# Output
print(f"\n--- Conversione di {euro:.2f} EUR ---")
print(f"  USD: {usd:.2f} $")
print(f"  GBP: {gbp:.2f} £")
print(f"  JPY: {jpy} ¥")
print(f"\nSupera i 1000 USD? {supera_mille}")
```

**Punti chiave:**
- I tassi di cambio sono in variabili con nomi in maiuscolo (`TC_USD`) — convenzione per indicare che sono **costanti** (valori che non cambiano).
- `int(euro * TC_JPY)` applica il casting per eliminare i decimali dello yen.
- `usd > 1000` è un'espressione booleana — il suo risultato è direttamente `True` o `False`.
- Il `bool` può essere mostrato direttamente con `print()` e le f-string.
</details>

## Riepilogo della lezione

| Tipo | Come si crea | A cosa serve | Casting |
|------|-------------|----------------|---------|
| `int` | `x = 5` | Contare, indicizzare, operare con interi | `int("5")`, `int(3.9)` → `3` |
| `float` | `x = 3.14` | Misure, percentuali, calcoli precisi | `float("3.14")`, `float(5)` |
| `str` | `x = "ciao"` | Testo, nomi, messaggi | `str(42)` → `"42"` |
| `bool` | `x = True` | Condizioni, confronti | `bool(0)` → `False` |

**Funzioni chiave di oggi:**
- `type(x)` — conoscere il tipo di una variabile
- `int()`, `float()`, `str()`, `bool()` — convertire i tipi
- `round(x, n)` — arrotondare a n decimali
- `abs(x)` — valore assoluto
- `len(testo)` — lunghezza di una stringa
- Metodi delle stringhe: `.upper()`, `.lower()`, `.strip()`, `.replace()`, `.split()`
