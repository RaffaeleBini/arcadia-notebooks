## 1. Il problema senza Git

Prima di vedere cos'è Git, vediamo il problema che risolve. Ti risulterà familiare.

### La cartella del caos

```
📁 mio_progetto/
├── analisi.py
├── analisi_v2.py
├── analisi_v2_finale.py
├── analisi_v2_finale_QUESTA.py
├── analisi_v2_finale_QUESTA_per_davvero.py
├── analisi_BUONA.py
├── analisi_BUONA_copia.py
└── analisi_da_consegnare_oggi.py
```

Qual è la versione corretta? Cosa è cambiato tra `_v2` e `_v2_finale`? Puoi tornare alla versione di tre giorni fa che funzionava?

### Il problema in un team

```
Lunedì:   Anna lavora su analisi.py    →  carica analisi_ANNA.py
Martedì:  Luigi lavora su analisi.py   →  carica analisi_LUIGI.py
Mercoledì: Chi fonde i due? Cosa succede ai cambiamenti che si sovrascrivono?
```

### I cinque problemi che Git risolve

| Problema | Senza Git | Con Git |
|----------|---------|--------|
| **Storico** | Nomi di file | Ogni cambiamento registrato con data, autore e messaggio |
| **Annullare** | Impossibile o manuale | `git revert` — torna a qualsiasi punto precedente |
| **Collaborazione** | Copie che si sovrascrivono | Ogni persona lavora sul proprio branch, Git fonde |
| **Contesto** | Perché ho cambiato questo? | Il messaggio di commit spiega il motivo |
| **Sicurezza** | Un solo punto di fallimento | Ogni sviluppatore ha una copia completa |

## 2. Cos'è Git

**Git** è un sistema di controllo di versione distribuito (**DVCS** — Distributed Version Control System). Lo ha creato Linus Torvalds nel 2005 per gestire lo sviluppo del kernel Linux.

### Come funziona Git internamente

Git non salva differenze riga per riga — salva **istantanee** (snapshot) dello stato completo del progetto a ogni commit.

```
Commit 1      Commit 2      Commit 3      Commit 4
──────────    ──────────    ──────────    ──────────
analisi.py    analisi.py*   analisi.py    analisi.py*
dati.csv      dati.csv      dati.csv*     dati.csv*
config.json   config.json   config.json   config.json*

* = file modificato in questo commit
  = stesso file del commit precedente (Git lo referenzia, non lo duplica)
```

Questo rende Git estremamente efficiente: non copia file che non sono cambiati.

### Distribuito vs centralizzato

```
Sistema CENTRALIZZATO (SVN, prima)   Sistema DISTRIBUITO (Git)
────────────────────────────────    ─────────────────────────────

       SERVER CENTRALE                    REPOSITORY REMOTO
           │                                   (GitHub)
    ┌──────┼──────┐                   ┌─────────┼─────────┐
  Anna    Luigi   Marta               Anna      Luigi     Marta
  (copia (copia  (copia              (copia    (copia    (copia
parziale)parziale)parziale)         COMPLETA) COMPLETA) COMPLETA)

Se il server cade → lavoro perso       Se GitHub cade → continuiamo a lavorare
Senza connessione → niente lavoro      Senza connessione → continuiamo con Git locale
```

### Vocabolario fondamentale

| Termine | Cos'è |
|---------|--------|
| **Repository** (`repo`) | Il progetto completo + tutto il suo storico di cambiamenti |
| **Commit** | Un'istantanea salvata del progetto in un dato momento |
| **Branch** (ramo) | Una linea di sviluppo indipendente |
| **Merge** | Fondere due branch |
| **Remote** | Una copia del repo su un server (GitHub, GitLab…) |
| **Clone** | Scaricare un repo remoto completo sulla tua macchina |
| **Push / Pull** | Inviare / ricevere cambiamenti al/dal remote |

### Dove viene salvata la configurazione

Git ha tre livelli di configurazione — ciascuno sovrascrive il precedente:

```
SISTEMA         →  /etc/gitconfig           (influisce su tutti gli utenti del SO)
GLOBALE         →  ~/.gitconfig             (influisce su tutti i tuoi progetti)
LOCALE          →  mio_progetto/.git/config (influisce solo su quel progetto)

Priorità: LOCALE > GLOBALE > SISTEMA
```

Puoi avere un'email personale in `--global` e un'email aziendale in `--local` dentro un progetto specifico.

## 4. Le tre zone di Git

Questo è il concetto più importante per capire come funziona Git. Ogni file nel tuo progetto può trovarsi in una di tre zone:

```
┌─────────────────────────────────────────────────────────────┐
│                      IL TUO PROGETTO                        │
│                                                             │
│  WORKING DIRECTORY    STAGING AREA       REPOSITORY        │
│  (directory di         (indice /           (.git/)          │
│   lavoro)              area di stage)                       │
│                                                             │
│  I file che tu         Ciò che includerai   Lo storico      │
│  modifichi con il      nel prossimo         permanente dei  │
│  tuo editor            commit               commit          │
│                                                             │
│       │                    │                    │           │
│       │   git add          │   git commit       │           │
│       │ ─────────────────► │ ─────────────────► │           │
│       │                    │                    │           │
│       │ ◄─────────────────────────────────────  │           │
│       │   git checkout / git restore            │           │
└─────────────────────────────────────────────────────────────┘
```

### Perché esiste lo Staging Area?

Può sembrare un passaggio extra inutile. Ma è una delle idee più geniali di Git:

```
Situazione senza staging area:
  Ho fatto 5 cambiamenti in 5 file → posso fare solo UN commit con tutto

Situazione con staging area:
  Ho fatto 5 cambiamenti in 5 file
  → Aggiungo allo stage i 2 file del bug che ho risolto → commit 1: "fix: correggere calcolo IVA"
  → Aggiungo allo stage i 3 file della nuova funzionalità → commit 2: "feat: aggiungere esportazione PDF"
  → Storico pulito, leggibile, con uno scopo
```

### Stati di un file

```
Untracked    →  Git non sa che esiste (file nuovo)
Tracked      →  Git lo conosce. Può essere:
  Unmodified →  uguale all'ultimo commit
  Modified   →  cambiato ma non aggiunto allo stage
  Staged     →  pronto per il prossimo commit
```

## 6. `.gitignore` — Cosa deve ignorare Git

Prima di fare il primo commit, configuriamo quali file **non devono mai** entrare nel repository. Il file `.gitignore` contiene pattern di nomi di file o directory che Git ignorerà completamente.

**Cosa si ignora abitualmente?**

| Categoria | Esempi | Perché |
|-----------|----------|--------|
| **Segreti** | `.env`, `*.key`, `segreti.*` | Non caricare mai credenziali o chiavi API |
| **Ambienti virtuali** | `venv/`, `.venv/`, `env/` | Pesanti e si rigenerano con `pip install` |
| **Cache di Python** | `__pycache__/`, `*.pyc` | File compilati, si generano da soli |
| **Dati grandi** | `*.csv`, `*.parquet`, `data/raw/` | Usare Git LFS o storage esterno |
| **File dell'editor** | `.vscode/`, `.idea/`, `*.swp` | Configurazione locale che non è del progetto |
| **Build e distribuzione** | `dist/`, `build/`, `*.egg-info/` | Si generano con il processo di build |

````python
# Creare il .gitignore prima del primo commit
gitignore_content = """# ─── Segreti e configurazione locale ──────────────────────
.env
*.env
segreti.*
*.key
*.pem

# ─── Python ────────────────────────────────────────────────
__pycache__/
*.py[cod]
*.pyo
*.pyd

# ─── Ambienti virtuali ─────────────────────────────────────
venv/
.venv/
env/
.env/

# ─── Jupyter ───────────────────────────────────────────────
.ipynb_checkpoints/

# ─── Editor ────────────────────────────────────────────────
.vscode/
.idea/
*.swp
*.swo
.DS_Store

# ─── Dati (grandi o sensibili) ─────────────────────────────
data/raw/
*.parquet
"""

Path(".gitignore").write_text(gitignore_content, encoding="utf-8")
print(".gitignore creato")
print()

# Ora git status non mostra segreti.env — è ignorato
!git status
````

````python
# Verificare esplicitamente che segreti.env sia ignorato
!git check-ignore -v segreti.env
````

### `git commit` — Salvare l'istantanea

Un commit è permanente e immutabile. Una volta fatto, quel momento del progetto rimane registrato per sempre nello storico.

**Anatomia di un buon messaggio di commit:**

```
tipo(ambito): descrizione breve all'imperativo (max 72 caratteri)

Corpo opzionale: spiega il PERCHÉ, non il COSA.
Il codice mostra già il cosa.

Closes #123
```

**Tipi più usati (Conventional Commits):**

| Tipo | Quando usarlo |
|------|---------------|
| `feat` | Nuova funzionalità |
| `fix` | Correzione di un bug |
| `docs` | Solo documentazione |
| `refactor` | Codice riorganizzato senza cambiare comportamento |
| `test` | Aggiungere o correggere test |
| `chore` | Manutenzione: dipendenze, configurazione |
| `init` | Commit iniziale del repository |

git restore file.py                 # scartare cambiamenti nel working directory
git restore --staged file.py        # togliere dallo stage senza perdere i cambiamenti
git revert <hash>                   # creare un commit che ne annulla un altro (sicuro)

from datetime import date

# ── COMMIT 2 ──────────────────────────────────────────────
Path("nota.py").write_text("""
from datetime import date

class Nota:
    _contatore = 0

    def __init__(self, titolo, contenuto):

        self.id             = Nota._contatore
        self.titolo         = titolo.strip()
        self.contenuto      = contenuto.strip()
        self.data_creazione = date.today()

    def __str__(self):
        return f"[{self.id}] {self.titolo} ({self.data_creazione})"
""".strip(), encoding="utf-8")

Path("gestore.py").write_text("""
from nota import Nota

class GestoreNote:
    def __init__(self):
        self._note = []

    def crea(self, titolo, contenuto):
        nota = Nota(titolo, contenuto)
        self._note.append(nota)
        return nota

    def elenca(self):
        if not self._note:
            print('Nessuna nota presente.')
            return
        for nota in self._note:
            print(nota)
""".strip(), encoding="utf-8")

gestore_v2 = Path("gestore.py").read_text(encoding="utf-8")

    def cerca(self, termine):

        return [

    def elimina(self, id_nota):
        nota = next((n for n in self._note if n.id == id_nota), None)

"""

**Punti chiave:**

- I messaggi seguono il formato `tipo: descrizione` — leggendo il log, chiunque capisce lo storico del progetto.
- Il commit di `docs` modifica solo il README — non mescola codice e documentazione.
</details>

### Esercizio 5 (sfida) — Audit del repository

**Consegna:** Scrivi una funzione Python `audit_repo(percorso_repo)` che analizzi il repository `sistema_note` e generi un report che includa:

1. Numero totale di commit.
2. Elenco dei commit con hash (7 caratteri), data, autore e messaggio.
3. File presenti nel repository e la loro dimensione.
4. I file ignorati da `.gitignore` che esistono nella directory.
5. Se ci sono cambiamenti non commitati (working directory o stage sporchi).

Usa `subprocess` per eseguire comandi Git da Python e analizzarne l'output.

**Esempio di output:**
```
═══════════════════════════════════════
  AUDIT — sistema_note
═══════════════════════════════════════
Commit totali: 6

STORICO:
  a1b2c3d  2025-06-20  Anna Rossi    docs: aggiungere sezione stato
  ...

FILE TRACCIATI: 5  (totale: 12.4 KB)
  README.md       1.2 KB
  ...

STATO: ✅ Working tree pulito
```

````python
# ✏️ IL TUO CODICE QUI
import subprocess
````

<details>
<summary>💡 Vedi soluzione — Esercizio 5</summary>

```python
import subprocess
from pathlib import Path

def git(comando, cwd):
    """Esegue un comando git e restituisce l'output come lista di righe."""
    risultato = subprocess.run(
        ["git"] + comando.split(),
        cwd=cwd,
        capture_output=True,
        text=True
    )
    return risultato.stdout.strip()


def audit_repo(percorso_repo):
    percorso = Path(percorso_repo).expanduser().resolve()
    nome = percorso.name

    print(f"\n{'═'*50}")
    print(f"  AUDIT — {nome}")
    print(f"  {percorso}")
    print(f"{'═'*50}")

    # ── 1. Numero di commit ────────────────────────────────
    totale_commit = git("rev-list --count HEAD", percorso)
    print(f"\nCommit totali: {totale_commit}")

    # ── 2. Storico ─────────────────────────────────────────
    print("\nSTORICO:")
    # Formato: hash|data|autore|messaggio
    log_raw = git('log --pretty=format:%h|%ad|%an|%s --date=short', percorso)
    if log_raw:
        for riga in log_raw.splitlines():
            parti = riga.split("|", 3)
            if len(parti) == 4:
                hash_, data, autore, messaggio = parti
                print(f"  {hash_:<8} {data}  {autore:<15} {messaggio}")

    # ── 3. File tracciati ──────────────────────────────────
    print("\nFILE TRACCIATI:")
    file_raw = git("ls-files", percorso)
    file_list = file_raw.splitlines() if file_raw else []
    totale_byte = 0
    for nome_file in file_list:
        percorso_file = percorso / nome_file
        if percorso_file.exists():
            dim = percorso_file.stat().st_size
            totale_byte += dim
            dim_str = f"{dim / 1024:.1f} KB" if dim > 1024 else f"{dim} B"
            print(f"  {nome_file:<25} {dim_str:>8}")
    print(f"  {'─'*35}")
    print(f"  {'TOTALE':<25} {totale_byte / 1024:.1f} KB ({len(file_list)} file)")

    # ── 4. File ignorati presenti ──────────────────────────
    print("\nFILE IGNORATI (presenti su disco):")
    ignorati_raw = git("ls-files --others --ignored --exclude-standard", percorso)
    ignorati = ignorati_raw.splitlines() if ignorati_raw else []
    if ignorati:
        for f in ignorati:
            print(f"  🚫 {f}")
    else:
        print("  (nessuno)")

    # ── 5. Stato del working tree ──────────────────────────
    print("\nSTATO:")
    status = git("status --porcelain", percorso)
    if not status:
        print("  ✅ Working tree pulito — tutto commitato")
    else:
        print("  ⚠️  Ci sono cambiamenti non commitati:")
        for riga in status.splitlines():
            codice_stato = riga[:2].strip()
            file_ = riga[3:]
            significato = {
                "M":  "Modificato",
                "A":  "Nuovo nello stage",
                "D":  "Eliminato",
                "??": "Non tracciato",
            }.get(codice_stato, codice_stato)
            print(f"  {significato:<20} {file_}")

    print(f"{'═'*50}")


# Eseguire l'audit
audit_repo("~/sistema_note")
```

**Punti chiave:**
- `subprocess.run` con `capture_output=True, text=True` è il modo corretto per catturare l'output di un comando shell in Python.
- `git log --pretty=format:...` permette di progettare esattamente quali campi vogliamo — senza analizzare testo libero.
- `git ls-files` elenca solo i file tracciati. `git ls-files --others --ignored --exclude-standard` elenca quelli ignorati presenti.
- `git status --porcelain` restituisce lo stato in formato leggibile da macchina — facile da analizzare, stabile tra le versioni di Git.
</details>

### Il ciclo di lavoro fondamentale

```
1. Modificare i file nel working directory
       │
       ▼
2. git status          → vedere cosa è cambiato
       │
       ▼
3. git add <file>      → spostare i cambiamenti rilevanti nello stage
       │
       ▼
4. git diff --staged   → rivedere esattamente cosa entrerà
       │
       ▼
5. git commit -m "tipo: descrizione chiara"
       │
       └──► tornare al passo 1
```

### Comandi che devi conoscere a memoria

```bash
git init                     # creare repo
git status                   # bussola — usalo sempre
git add .                    # aggiungere tutto allo stage
git commit -m "messaggio"    # salvare l'istantanea
git log --oneline            # vedere lo storico
git diff                     # vedere i cambiamenti non aggiunti
git diff --staged            # vedere i cambiamenti nello stage
```

### Buone pratiche fin dal primo giorno

| Pratica | Perché |
|----------|---------|
| Commit piccoli e frequenti | Più facili da capire e annullare |
| Un commit = uno scopo | Lo storico racconta una storia |
| `.gitignore` prima del primo commit | Niente di sensibile entra per errore |
| `git status` prima di `git commit` | Rivedi cosa stai salvando |
| Messaggi all'imperativo | Convenzione universale: `"aggiungere"`, non `"aggiunto"` |
| Non commitare mai credenziali | Una volta nello storico, è molto difficile cancellarle |
