## 2. Struttura del filesystem Linux

In Linux (e macOS), tutto dipende da un unico punto: la **radice `/`**.

```
/                          ← radice (root)
├── bin/                   ← eseguibili essenziali del sistema (ls, cp, mv…)
├── etc/                   ← file di configurazione del sistema
├── home/                  ← directory personali degli utenti
│   ├── ana/               ← directory personale di ana  (~ per ana)
│   └── carlos/            ← directory personale di carlos
├── tmp/                   ← file temporanei (vengono eliminati al riavvio)
├── usr/                   ← programmi installati dall'utente
│   └── local/
│       └── bin/           ← programmi installati manualmente
├── var/                   ← dati variabili: log, database…
│   └── log/               ← log di sistema
└── dev/                   ← dispositivi (disco, USB, terminale…)
```

**Su macOS** la struttura è simile ma con alcune differenze:
- Le directory personali si trovano in `/Users/`, non in `/home/`.
- I programmi installati con Homebrew vanno in `/opt/homebrew/` (Apple Silicon) o `/usr/local/` (Intel).

**`~` (tilde)** è una scorciatoia per la tua directory personale: `~/proyectos` equivale a `/home/tuo-utente/proyectos`.

## 3. Percorsi assoluti e relativi

Un **percorso** è l'indirizzo di un file o directory nel sistema.

```
PERCORSO ASSOLUTO                  PERCORSO RELATIVO
─────────────────────────────      ────────────────────────────────
Inizia sempre con /                Inizia da dove ti trovi ora
/home/ana/proyectos/curso          proyectos/curso
Funziona da qualsiasi posto        Dipende dalla directory attuale
Più lungo da scrivere              Più corto
```

**Scorciatoie di percorso:**

| Scorciatoia | Significa |
|-------|-----------|
| `.`   | La directory attuale |
| `..`  | La directory padre (un livello sopra) |
| `~`   | La tua directory personale (`/home/utente`) |
| `-`   | La directory precedente (dove ti trovavi prima) |

```bash
# Esempi con le scorciatoie:
cd ..          # salire di un livello
cd ../..       # salire di due livelli
cd ~/proyectos # andare in proyectos nella tua home, da qualsiasi posto
cd -           # tornare alla directory precedente
```

## 5. `ls` — Cosa c'è qui?

`ls` (*list*) mostra il contenuto di una directory.

```bash
ls                    # contenuto della directory attuale
ls /percorso/directory # contenuto di un'altra directory
ls -l                 # formato lungo (permessi, dimensione, data)
ls -a                 # includere file nascosti (iniziano con .)
ls -la                # lungo + nascosti (i flag si combinano)
ls -lh                # lungo + dimensioni leggibili (KB, MB, GB)
ls -lt                # ordinare per data di modifica
ls -lS                # ordinare per dimensione
ls *.py               # solo file .py (glob)
```

````python
import os, shutil
from pathlib import Path

# Creare una directory di demo con struttura
DEMO = Path.home() / "linux_demo"
if DEMO.exists(): shutil.rmtree(DEMO)

# Creare struttura
(DEMO / "proyectos" / "curso_ia").mkdir(parents=True)
(DEMO / "proyectos" / "trabajo").mkdir(parents=True)
(DEMO / "datos").mkdir()
(DEMO / ".config").mkdir()  # cartella nascosta

(DEMO / "notas.txt").write_text("Le mie note\n", encoding="utf-8")
(DEMO / "script.py").write_text("print('ciao')\n", encoding="utf-8")
(DEMO / "datos" / "dataset.csv").write_text("id,valor\n1,100\n", encoding="utf-8")
(DEMO / ".config" / "settings.json").write_text("{}", encoding="utf-8")

os.chdir(DEMO)
print(f"Directory di demo: {DEMO}")
````

````python
# ls di base
print("=== ls (base) ===")
!ls

print("\n=== ls -l (formato lungo) ===")
!ls -l

print("\n=== ls -la (con file nascosti) ===")
!ls -la
````

````python
# Capire il formato lungo di ls -l
print("""
drwxr-xr-x  3  ana  staff   96  gen 15 10:30  proyectos
│└─┬──┘└─┬┘ │  │    │       │   │              └─ nome
│  │     │  │  │    │       │   └─ data di modifica
│  │     │  │  │    │       └─ dimensione in byte
│  │     │  │  │    └─ gruppo proprietario
│  │     │  │  └─ utente proprietario
│  │     │  └─ numero di collegamenti
│  │     └─ permessi del gruppo
│  └─ permessi dell'utente proprietario
└─ tipo: d=directory, -=file, l=collegamento simbolico
""")

print("\n=== ls -lh (dimensioni leggibili) ===")
!ls -lh datos/
````

## 6. `cd` — Navigare tra directory

`cd` (*change directory*) ti sposta in un'altra directory.

```bash
cd /percorso/assoluto     # andare a un percorso assoluto
cd nome_cartella     # andare in una sottocartella (relativo)
cd ..                 # salire di un livello
cd ../altra_cartella    # salire ed entrare in un'altra
cd ~                  # andare alla home
cd                    # va anche alla home (senza argomenti)
cd -                  # tornare alla directory precedente
```

````python
# Nota: in Jupyter, !cd non persiste tra le celle.
# Usa os.chdir() per spostare la directory di lavoro di Python,
# oppure usa %%bash per eseguire più righe nella stessa shell.

print("Dimostrazione di navigazione:")
!pwd

print("\nEntrare in proyectos:")
os.chdir(DEMO / "proyectos")
!pwd

print("\nSalire di un livello (..):")
os.chdir(DEMO)
!pwd

print("\nPercorso relativo composto (proyectos/curso_ia):")
os.chdir(DEMO / "proyectos" / "curso_ia")
!pwd

print("\nTornare al demo con ~ (home relativo):")
os.chdir(DEMO)
!pwd
````

````python
%%bash
# Con %%bash, tutte le righe vengono eseguite nella stessa shell
# quindi cd persiste all'interno di questo blocco

cd ~/linux_demo
echo "1. Siamo in: $(pwd)"

cd proyectos
echo "2. Entriamo in: $(pwd)"

cd curso_ia
echo "3. Entriamo in: $(pwd)"

cd ../..
echo "4. Saliamo di due livelli: $(pwd)"

cd -
echo "5. Torniamo alla precedente: $(pwd)"
````

## 7. `mkdir` e `rmdir` — Creare ed eliminare directory


**`-p`** (*parents*) è il flag più utile: crea tutte le directory intermedie che non esistono.

## 8. `tree` — Vedere la struttura completa

`tree` mostra l'albero delle directory in modo visuale. Molto utile per capire la struttura di un progetto.

```bash
tree                  # albero completo dalla directory attuale
tree -L 2             # solo fino a 2 livelli di profondità
tree -a               # includere file nascosti
tree -d               # solo directory (senza file)
tree --noreport       # senza la riga di riepilogo finale
```

> Se `tree` non è installato: `brew install tree` (macOS) o `apt install tree` (Linux).

````python
# tree (se installato)
!tree -L 2 2>/dev/null || echo "tree non installato — uso l'alternativa Python"

# Alternativa Python quando tree non è disponibile
def mostrar_arbol(ruta, prefijo="", max_nivel=3, nivel=0):
    if nivel > max_nivel:
        return
    ruta = Path(ruta)
    contenido = sorted(ruta.iterdir(), key=lambda p: (p.is_file(), p.name))
    for i, item in enumerate(contenido):
        es_ultimo = i == len(contenido) - 1
        conector = "└── " if es_ultimo else "├── "
        tipo = "/" if item.is_dir() else ""
        print(f"{prefijo}{conector}{item.name}{tipo}")
        if item.is_dir():
            extension = "    " if es_ultimo else "│   "
            mostrar_arbol(item, prefijo + extension, max_nivel, nivel + 1)

print(f"\n{DEMO.name}/")
mostrar_arbol(DEMO)
````

### Esercizio 5 (sfida) — Esploratore di progetti

**Consegna:** Scrivi la funzione Python `explorar_proyecto(ruta)` che generi un report completo di una directory:

- Numero totale di file e directory.
- Distribuzione per estensione (quanti `.py`, `.txt`, `.csv`, ecc.).
- I cinque file più grandi con la loro dimensione leggibile (KB/MB).
- File nascosti trovati.
- Profondità massima dell'albero.

Provala con `~/ml_proyecto`.

````python
# ✏️ IL TUO CODICE QUI
from pathlib import Path
````

<details>
<summary>💡 Vedi soluzione — Esercizio 5</summary>

```python
from pathlib import Path
from collections import Counter

def formato_bytes(n):
    for unidad in ("B", "KB", "MB", "GB"):
        if n < 1024:
            return f"{n:.1f} {unidad}"
        n /= 1024
    return f"{n:.1f} TB"

def explorar_proyecto(ruta):
    ruta = Path(ruta).expanduser().resolve()

    archivos = [p for p in ruta.rglob("*") if p.is_file()]
    directorios = [p for p in ruta.rglob("*") if p.is_dir()]
    ocultos = [p for p in archivos if p.name.startswith(".")]

    extensiones = Counter(p.suffix.lower() or "(senza estensione)" for p in archivos)

    mas_grandes = sorted(archivos, key=lambda p: p.stat().st_size, reverse=True)[:5]

    def profundidad(p):
        return len(p.relative_to(ruta).parts)

    prof_max = max((profundidad(p) for p in archivos), default=0)

    print(f"\n{'═'*55}")
    print(f"  REPORT: {ruta.name}")
    print(f"{'═'*55}")
    print(f"  File:         {len(archivos)}")
    print(f"  Directory:    {len(directorios)}")
    print(f"  Profondità:   {prof_max} livelli")

    print(f"\n  ESTENSIONI:")
    for ext, count in extensiones.most_common():
        print(f"    {ext:<20} {count:>3} file")

    print(f"\n  5 FILE PIÙ GRANDI:")
    for p in mas_grandes:
        tam = formato_bytes(p.stat().st_size)
        rel = p.relative_to(ruta)
        print(f"    {tam:>10}  {rel}")

    print(f"\n  FILE NASCOSTI ({len(ocultos)}):")
    for p in ocultos:
        print(f"    {p.relative_to(ruta)}")
    if not ocultos:
        print("    Nessuno")

    print(f"{'═'*55}")


explorar_proyecto("~/ml_proyecto")
```
