## ¿Por qué importan los tipos de datos?

Imagina que tienes dos variables:

```python
a = "3"
b = "5"
print(a + b)   # → "35"  ← concatena texto
```

vs.

```python
a = 3
b = 5
print(a + b)   # → 8  ← suma números
```

El operador `+` hace cosas completamente distintas dependiendo del **tipo** de los datos. Entender los tipos es esencial para evitar errores difíciles de detectar.

````python
# Ejemplo: el mismo operador, resultados muy diferentes
a_texto  = "3"
b_texto  = "5"
a_numero = 3
b_numero = 5

print("Texto + texto:   ", a_texto  + b_texto)   # concatena
print("Número + número: ", a_numero + b_numero)   # suma
````

## 1. `int` — Números enteros

Un `int` es cualquier número sin parte decimal: positivo, negativo o cero.

```python
edad        = 28
temperatura = -5
cero        = 0
poblacion   = 8_000_000_000   # puedes usar _ como separador de miles
```

### ¿Hasta qué número puede llegar un `int`?

En Python 3, los enteros **no tienen límite de tamaño**. Puedes representar números astronómicamente grandes sin ningún problema. Esto no es así en todos los lenguajes.

### Operaciones específicas de enteros

| Operador | Qué hace | Ejemplo | Resultado |
|----------|----------|---------|----------|
| `//` | División entera (sin decimales) | `17 // 5` | `3` |
| `%` | Módulo (resto de la división) | `17 % 5` | `2` |
| `**` | Potencia | `2 ** 10` | `1024` |

````python
# Ejemplos con int
edad       = 28
año_actual = 2025
poblacion  = 8_000_000_000   # el guion bajo no afecta al valor, solo a la lectura

print(type(edad))
print(f"Edad: {edad}")
print(f"Año de nacimiento: {año_actual - edad}")
print(f"Población mundial: {poblacion:,}")   # :, formatea con comas como separador de miles
````

````python
# Operaciones exclusivas de enteros
a = 17
b = 5

print(f"{a} // {b} = {a // b}")   # cuántas veces cabe b en a
print(f"{a} %  {b} = {a % b}")    # lo que sobra
print(f"2 ** 10 = {2 ** 10}")      # 2 elevado a 10

# Curiosidad: int de tamaño arbitrario
numero_gigante = 2 ** 100
print(f"2 elevado a 100 = {numero_gigante}")
````

## 2. `float` — Números decimales

Un `float` es un número con parte decimal. Internamente, Python los representa en notación de punto flotante (de ahí el nombre).

```python
precio   = 19.99
pi       = 3.14159
negativo = -0.5
ciencia  = 1.5e6    # notación científica: 1.5 × 10⁶ = 1 500 000
```

### ⚠️ El problema de la precisión flotante

Los `float` tienen un límite de precisión. Esto puede provocar resultados inesperados:

```python
0.1 + 0.2   # → 0.30000000000000004  (¡no es 0.3!)
```

Esto no es un bug de Python — es una limitación de cómo los ordenadores representan decimales en binario. Para cálculos financieros precisos, se usa la librería `decimal`.

````python
# Ejemplos con float
precio  = 19.99
pi      = 3.14159
ciencia = 1.5e6     # notación científica

print(type(precio))
print(f"Precio:   {precio}")
print(f"Pi:       {pi}")
print(f"Ciencia:  {ciencia}")        # muestra 1500000.0
print(f"Ciencia:  {ciencia:,.0f}")   # formateado con separador de miles
````

````python
# El problema de precisión de los floats
resultado = 0.1 + 0.2
print(f"0.1 + 0.2 = {resultado}")             # muestra el problema
print(f"0.1 + 0.2 = {resultado:.2f}")         # redondeado a 2 decimales

# Comparación: nunca compares floats con ==
print(0.1 + 0.2 == 0.3)                       # False ← trampa clásica
print(round(0.1 + 0.2, 2) == 0.3)             # True ← forma correcta

# round() redondea a N decimales
print(round(3.14159, 2))                       # 3.14
````

### Funciones matemáticas útiles con números

````python
import math

x = -7.8

print(f"abs({x})           = {abs(x)}")           # valor absoluto
print(f"round({x})         = {round(x)}")         # redondeo al entero más cercano
print(f"math.floor({x})    = {math.floor(x)}")    # redondeo hacia abajo
print(f"math.ceil({x})     = {math.ceil(x)}")     # redondeo hacia arriba
print(f"math.sqrt(144)     = {math.sqrt(144)}")   # raíz cuadrada
print(f"math.log10(1000)   = {math.log10(1000)}") # logaritmo base 10
````

## 3. `str` — Cadenas de texto

Un `str` (string) es una secuencia de caracteres. Se escribe entre comillas simples `'...'` o dobles `"..."`. Ambas son equivalentes.

```python
nombre   = "Ana"
mensaje  = 'Hola, mundo'
vacia    = ""            # string vacío
multilínea = """Este texto
ocupa varias
líneas"""                # tres comillas para varias líneas
```

### Los strings son secuencias

Cada carácter de un string tiene una **posición** (índice), empezando desde 0:

```
  texto = "Python"
            P  y  t  h  o  n
  índice:   0  1  2  3  4  5
  negativo:-6 -5 -4 -3 -2 -1
```

Puedes acceder a un carácter con `texto[índice]` y a una parte con `texto[inicio:fin]` (**slicing**).

````python
texto = "Python"

# Acceso por índice
print(f"Primer carácter:  {texto[0]}")    # P
print(f"Último carácter:  {texto[-1]}")   # n
print(f"Tercer carácter:  {texto[2]}")    # t

# Slicing: texto[inicio:fin]  (fin no incluido)
print(f"Primeros 3:       {texto[0:3]}")  # Pyt
print(f"Desde el 3:       {texto[3:]}")   # hon
print(f"Últimos 3:        {texto[-3:]}")  # hon
print(f"Longitud:         {len(texto)}")  # 6
````

### Métodos de los strings

Los strings tienen muchos **métodos** — funciones que pertenecen al propio string y se llaman con un punto.

| Método | Qué hace | Ejemplo |
|--------|----------|--------|
| `.upper()` | Convierte a mayúsculas | `"hola".upper()` → `"HOLA"` |
| `.lower()` | Convierte a minúsculas | `"HOLA".lower()` → `"hola"` |
| `.strip()` | Elimina espacios al inicio y al final | `" hola ".strip()` → `"hola"` |
| `.replace(a, b)` | Reemplaza `a` por `b` | `"hola mundo".replace("mundo", "Python")` |
| `.split(sep)` | Divide el string en una lista | `"a,b,c".split(",")` → `['a', 'b', 'c']` |
| `.startswith(x)` | ¿Empieza por `x`? | `"Python".startswith("Py")` → `True` |
| `.count(x)` | Cuántas veces aparece `x` | `"banana".count("a")` → `3` |
| `.find(x)` | Posición de `x` (-1 si no existe) | `"Python".find("th")` → `2` |

````python
frase = "  Hola, Mundo!  "

print(frase.strip())           # elimina espacios
print(frase.strip().upper())   # encadenamos métodos
print(frase.strip().lower())
print(frase.strip().replace("Mundo", "Python"))
print(frase.strip().split(", "))  # divide por ", "

# Contar y buscar
texto = "banana"
print(f"La letra 'a' aparece {texto.count('a')} veces")
print(f"La posición de 'nan' es {texto.find('nan')}")
````

````python
# Operaciones con strings
saludo  = "Hola"
nombre  = "Ana"

# Concatenación con +
mensaje = saludo + ", " + nombre + "!"
print(mensaje)

# Repetición con *
separador = "-" * 30
print(separador)

# f-string (la forma más recomendada)
print(f"{saludo}, {nombre}!")

# Verificar si un texto contiene algo (operador 'in')
print("Ana" in mensaje)    # True
print("Pedro" in mensaje)  # False
````

### Caracteres especiales (secuencias de escape)

Algunos caracteres tienen significado especial dentro de un string:

| Secuencia | Qué produce |

| `\n` | Salto de línea |

| `\'` | Comilla simple literal |

````python
print("Línea 1\nLínea 2\nLínea 3")
print("-" * 20)
print("Columna 1\tColumna 2\tColumna 3")
print("-" * 20)
print("Ella dijo: \"Hola\"")
````

## 4. `bool` — Valores lógicos

Un `bool` solo puede tener dos valores: `True` (verdadero) o `False` (falso). Son la base de toda lógica en programación.

> ⚠️ Importante: en Python `True` y `False` se escriben con la primera letra en mayúscula.

### ¿De dónde vienen los booleanos?

La mayoría de las veces no los escribes directamente — son el resultado de una **comparación** o una **condición**:

````python
# Operadores de comparación → siempre devuelven bool
print(5 > 3)      # True
print(5 < 3)      # False
print(5 == 5)     # True  (== es igualdad; = es asignación)
print(5 != 3)     # True  (!= es diferente)
print(5 >= 5)     # True
print(5 <= 4)     # False

# Con variables
edad = 20
es_mayor_de_edad = edad >= 18
print(f"¿Es mayor de edad? {es_mayor_de_edad}")
````

### Operadores lógicos: `and`, `or`, `not`

Combinan condiciones booleanas:

| Operador | Significado | Resultado |
|----------|-------------|----------|
| `A and B` | Las dos deben ser `True` | `True` solo si A y B son `True` |
| `A or B` | Al menos una debe ser `True` | `False` solo si A y B son `False` |
| `not A` | Invierte el valor | `not True` → `False` |

```
Tabla de verdad de AND:
  True  and True   → True
  True  and False  → False
  False and True   → False
  False and False  → False

Tabla de verdad de OR:
  True  or True   → True
  True  or False  → True
  False or True   → True
  False or False  → False
```

````python
edad       = 20
tiene_dni  = True

# and: las dos condiciones deben ser verdaderas
puede_votar = edad >= 18 and tiene_dni
print(f"¿Puede votar? {puede_votar}")

# or: con que una sea verdadera es suficiente
es_estudiante = False
es_jubilado   = True
tiene_descuento = es_estudiante or es_jubilado
print(f"¿Tiene descuento? {tiene_descuento}")

# not: invierte
print(f"¿No es estudiante? {not es_estudiante}")
````

### Valores "truthy" y "falsy"

En Python, cualquier valor puede usarse en un contexto booleano. Algunos valores se consideran `False` automáticamente:

| Valor | ¿Es `False`? |
|-------|-------------|
| `0` (entero cero) | Sí |
| `0.0` (float cero) | Sí |
| `""` (string vacío) | Sí |
| `None` (ausencia de valor) | Sí |
| Cualquier otro valor | No (es `True`) |

````python
# bool() convierte cualquier valor a booleano
print(bool(0))       # False
print(bool(0.0))     # False
print(bool(""))      # False
print(bool(None))    # False

print("---")

print(bool(1))       # True
print(bool(-5))      # True  (cualquier número no-cero)
print(bool("hola"))  # True  (cualquier string no vacío)
print(bool(0.001))   # True
````

## 5. Casting — Convertir tipos

El **casting** (o conversión de tipos) es transformar un valor de un tipo a otro.

En Python se hace con las funciones de conversión:

| Función | Convierte a | Ejemplo |
|---------|------------|--------|
| `int(x)` | Entero | `int("42")` → `42` |
| `float(x)` | Decimal | `float("3.14")` → `3.14` |
| `str(x)` | Texto | `str(100)` → `"100"` |
| `bool(x)` | Booleano | `bool(0)` → `False` |

### ¿Cuándo lo necesitas?

El caso más frecuente: `input()` **siempre devuelve un string**, aunque el usuario escriba un número. Debes convertirlo antes de operar.

```python
dato = input("Escribe un número: ")   # dato es "42" (str)
numero = int(dato)                    # ahora es 42 (int)
```

````python
# int() — convierte a entero
print(int("42"))       # de str a int
print(int(3.9))        # de float a int (trunca, NO redondea)
print(int(True))       # True → 1
print(int(False))      # False → 0

print("---")

# float() — convierte a decimal
print(float("3.14"))   # de str a float
print(float(10))       # de int a float
print(float("1e3"))    # notación científica en string

print("---")

# str() — convierte a texto
print(str(42))         # de int a str
print(str(3.14))       # de float a str
print(str(True))       # de bool a str
````

````python
# ⚠️ Conversiones que fallan
# Descomenta cada línea para ver el error

# int("hola")       # ValueError: no se puede convertir texto no numérico a int
# int("3.14")       # ValueError: no se puede convertir un float-string directamente a int
# float("abc")      # ValueError

# Solución para "3.14" → int: pasar primero por float
resultado = int(float("3.14"))
print(f"int(float('3.14')) = {resultado}")
````

````python
# Python hace algunas conversiones automáticas (implícitas)
# int + float → float  (Python elige el tipo más preciso)
resultado = 3 + 1.5
print(resultado)          # 4.5
print(type(resultado))    # float

# bool en operaciones numéricas
print(True + True)        # 2  (True = 1)
print(True + 5)           # 6
print(False * 100)        # 0  (False = 0)
````

### Resumen visual de conversiones

```
         int()
  str ──────────────▶ int
  float ────(trunca)─▶ int
  bool ─────(0 o 1)──▶ int

         float()
  str ──────────────▶ float
  int ──────────────▶ float

         str()
  int   ────────────▶ str
  float ────────────▶ str
  bool  ────────────▶ str  ("True" / "False")

         bool()
  0, 0.0, "", None  ──▶ False
  cualquier otro    ──▶ True
```

### Ejercicio 1 — Conversor de temperatura

**Enunciado:** Escribe un programa que pida una temperatura en **grados Celsius** y la convierta a:
- **Fahrenheit:** `F = C × 9/5 + 32`
- **Kelvin:** `K = C + 273.15`

**Ejemplo de salida esperada:**
```
Introduce la temperatura en Celsius: 100

100.00 °C equivale a:
  212.00 °F (Fahrenheit)
  373.15 K  (Kelvin)
```

````python
# ✏️ TU CÓDIGO AQUÍ
````

### Ejercicio 4 — Conversor de tiempo

**Enunciado:** Escribe un programa que pida un número de **segundos** (entero) y lo descomponga en:
- Días
- Horas restantes
- Minutos restantes
- Segundos restantes

**Pista:** usa división entera `//` y módulo `%`.

**Ejemplo de salida esperada:**
```
Introduce el número de segundos: 90061

90061 segundos equivale a:
  1 día(s)
  1 hora(s)
  1 minuto(s)
  1 segundo(s)
```

````python
# ✏️ TU CÓDIGO AQUÍ
# Pista:
#   1 minuto  = 60 segundos
#   1 hora    = 3600 segundos
#   1 día     = 86400 segundos
#
# Estrategia: ir de la unidad más grande a la más pequeña
#   días     = total // 86400
#   resto    = total % 86400   ← los segundos que sobran tras sacar los días
#   horas    = resto // 3600   ← y así sucesivamente
````

<details>
<summary>💡 Ver solución — Ejercicio 4</summary>

```python
total = int(input("Introduce el número de segundos: "))

dias    = total // 86400
resto   = total  % 86400

horas   = resto  // 3600
resto   = resto   % 3600

minutos = resto  // 60
segundos = resto  % 60

print(f"\n{total} segundos equivale a:")
print(f"  {dias} día(s)")
print(f"  {horas} hora(s)")
print(f"  {minutos} minuto(s)")
print(f"  {segundos} segundo(s)")
```

**Puntos clave:**
- La estrategia es siempre la misma: `//` para obtener la cantidad entera de la unidad grande, `%` para obtener el resto que pasamos a la siguiente unidad más pequeña.
- Reutilizamos la variable `resto` en cada paso — es válido porque ya no necesitamos el valor anterior.
- `int()` es necesario porque `input()` devuelve un string y los operadores `//` y `%` necesitan enteros.
</details>

### Ejercicio 5 (reto) — Conversor completo con validación de tipo

**Enunciado:** Crea un conversor de moneda que:
1. Pida una cantidad en **euros** (`float`).
2. Use estos tipos de cambio:
   - `1 EUR = 1.08 USD`
   - `1 EUR = 0.86 GBP`
   - `1 EUR = 162.5 JPY`
3. Muestre los resultados con el formato de cada moneda:
   - USD y GBP con 2 decimales
   - JPY como número entero (sin decimales, porque el yen no tiene céntimos)
4. Muestra también si la cantidad en USD supera los 1 000 dólares usando un `bool`.

**Ejemplo de salida esperada:**
```
Introduce la cantidad en euros: 950

--- Conversión de 950.00 EUR ---
  USD: 1026.00 $
  GBP: 817.00 £
  JPY: 154375 ¥

¿Supera los 1000 USD? True
```

````python
# ✏️ TU CÓDIGO AQUÍ
````

<details>
<summary>💡 Ver solución — Ejercicio 5</summary>

```python
euros = float(input("Introduce la cantidad en euros: "))

# Tipos de cambio
TC_USD = 1.08
TC_GBP = 0.86
TC_JPY = 162.5

# Conversiones
usd = euros * TC_USD
gbp = euros * TC_GBP
jpy = int(euros * TC_JPY)   # el yen es entero

# ¿Supera los 1000 USD?
supera_mil = usd > 1000     # esto es un bool

# Salida
print(f"\n--- Conversión de {euros:.2f} EUR ---")
print(f"  USD: {usd:.2f} $")
print(f"  GBP: {gbp:.2f} £")
print(f"  JPY: {jpy} ¥")
print(f"\n¿Supera los 1000 USD? {supera_mil}")
```

**Puntos clave:**
- Los tipos de cambio están en variables con nombres en mayúsculas (`TC_USD`) — convención para indicar que son **constantes** (valores que no cambian).
- `int(euros * TC_JPY)` aplica casting para eliminar los decimales del yen.
- `usd > 1000` es una expresión booleana — su resultado es `True` o `False` directamente.
- El `bool` se puede mostrar directamente con `print()` y f-strings.
</details>

|------|-------------|----------------|---------|
| `int` | `x = 5` | Contar, indexar, operar con enteros | `int("5")`, `int(3.9)` → `3` |
| `float` | `x = 3.14` | Medidas, porcentajes, cálculos precisos | `float("3.14")`, `float(5)` |
| `str` | `x = "hola"` | Texto, nombres, mensajes | `str(42)` → `"42"` |
| `bool` | `x = True` | Condiciones, comparaciones | `bool(0)` → `False` |

**Funciones clave de hoy:**
- `type(x)` — conocer el tipo de una variable
- `int()`, `float()`, `str()`, `bool()` — convertir tipos
- `round(x, n)` — redondear a n decimales
- `abs(x)` — valor absoluto
- `len(texto)` — longitud de un string
- Métodos de string: `.upper()`, `.lower()`, `.strip()`, `.replace()`, `.split()`
