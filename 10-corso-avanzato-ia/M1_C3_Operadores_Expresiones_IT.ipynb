## 1. Operatori aritmetici

Sono gli operatori matematici classici. Li abbiamo già visti nella Lezione 1, qui li approfondiamo.

| Operatore | Nome | Esempio | Risultato |
|----------|--------|---------|----------|
| `+` | Somma | `7 + 3` | `10` |
| `-` | Sottrazione | `7 - 3` | `4` |
| `*` | Moltiplicazione | `7 * 3` | `21` |
| `/` | Divisione reale | `7 / 2` | `3.5` |
| `//` | Divisione intera | `7 // 2` | `3` |
| `%` | Modulo (resto) | `7 % 2` | `1` |
| `**` | Potenza | `2 ** 8` | `256` |
| `-x` | Negazione unaria | `-7` | `-7` |

### Relazione tra `/`, `//` e `%`

Sono legati da questa formula:
```
a = (a // b) * b + (a % b)
```
Esempio: `17 = (17 // 5) * 5 + (17 % 5)` → `17 = 3 * 5 + 2` → `17 = 17` ✓

````python
a, b = 17, 5   # assegnazione multipla: a = 17, b = 5 in una sola riga

print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} / {b}  = {a / b}")    # sempre float
print(f"{a} // {b} = {a // b}")   # quoziente intero
print(f"{a} % {b}  = {a % b}")    # resto
print(f"{a} ** {b} = {a ** b}")   # potenza

# Verifichiamo la relazione tra //, % e l'originale
verifica = (a // b) * b + (a % b)
print(f"\nVerifica: ({a}//{b})*{b} + ({a}%{b}) = {verifica} == {a}")
````

### 1.1 Precedenza degli operatori (ordine di valutazione)

Quando un'espressione ha più operatori, Python li valuta in un ordine preciso — proprio come in matematica:

| Priorità | Operatore/i | Nome |
|-----------|-------------|--------|
| 1 (massima) | `()` | Parentesi |
| 2 | `**` | Potenza |
| 3 | `-x` | Negazione unaria |
| 4 | `*`, `/`, `//`, `%` | Moltiplicazione e divisione |
| 5 (minima) | `+`, `-` | Somma e sottrazione |

> **Regola d'oro:** in caso di dubbio, usa le **parentesi**. Rendono il codice più leggibile ed evitano errori.

````python
# Senza parentesi: Python applica le regole di precedenza
print(2 + 3 * 4)       # 14, non 20 (prima * poi +)
print(10 - 2 ** 3)     # 2,  non 512 (prima ** poi -)
print(10 / 2 + 3)      # 8.0 (prima / poi +)

print("---")

# Con parentesi: controlliamo l'ordine
print((2 + 3) * 4)     # 20
print((10 - 2) ** 3)   # 512
print(10 / (2 + 3))    # 2.0

print("---")

# Esempio reale: formula dell'interesse composto   C * (1 + r)^t
capitale = 1000
tasso    = 0.05    # 5% annuo
anni     = 10

# SBAGLIATO (senza parentesi, la precedenza cambia il risultato)
sbagliato = capitale * 1 + tasso ** anni
# CORRETTO
corretto  = capitale * (1 + tasso) ** anni

print(f"Senza parentesi (SBAGLIATO): {sbagliato:.2f} €")
print(f"Con parentesi (CORRETTO): {corretto:.2f} €")
````

### 1.2 Operatori di assegnazione composta

Sono scorciatoie per modificare una variabile applicandole un'operazione:

| Operatore | Equivale a | Esempio |
|----------|-----------|--------|
| `x += 5` | `x = x + 5` | Aggiunge 5 a x |
| `x -= 3` | `x = x - 3` | Sottrae 3 a x |
| `x *= 2` | `x = x * 2` | Moltiplica x per 2 |
| `x /= 4` | `x = x / 4` | Divide x per 4 |
| `x //= 3` | `x = x // 3` | Divisione intera di x per 3 |
| `x %= 7` | `x = x % 7` | Resto di x diviso 7 |
| `x **= 2` | `x = x ** 2` | x elevato al quadrato |

````python
punti = 100
print(f"Inizio:            {punti}")

punti += 50      # guadagni 50 punti
print(f"Dopo aver guadagnato 50: {punti}")

punti -= 30      # perdi 30 punti
print(f"Dopo aver perso 30:      {punti}")

punti *= 2       # punti doppi
print(f"Punti doppi:       {punti}")

punti //= 3      # dividi tra 3 squadre
print(f"Divisi in 3:       {punti}")
````

## 2. Operatori di confronto

Confrontano due valori e **restituiscono sempre un booleano** (`True` o `False`).

| Operatore | Significato | Esempio | Risultato |
|----------|------------|---------|----------|
| `==` | Uguale a | `5 == 5` | `True` |
| `!=` | Diverso da | `5 != 3` | `True` |
| `>` | Maggiore di | `5 > 3` | `True` |
| `<` | Minore di | `5 < 3` | `False` |
| `>=` | Maggiore o uguale | `5 >= 5` | `True` |
| `<=` | Minore o uguale | `5 <= 4` | `False` |

> ⚠️ **Errore classico:** confondere `=` (assegnazione) con `==` (confronto).
> ```python
> x = 5      # assegna il valore 5 a x
> x == 5     # chiede se x vale 5 → True
> ```

````python
a = 10
b = 7

print(f"{a} == {b}  →  {a == b}")
print(f"{a} != {b}  →  {a != b}")
print(f"{a} >  {b}  →  {a > b}")
print(f"{a} <  {b}  →  {a < b}")
print(f"{a} >= {b}  →  {a >= b}")
print(f"{a} <= {b}  →  {a <= b}")
````

````python
# Confrontare stringhe: Python confronta lettera per lettera in ordine alfabetico (Unicode)
print("apple" == "apple")   # True
print("apple" != "banana")  # True
print("apple" < "banana")   # True  (a < b nell'alfabeto)
print("Zebra" < "apple")    # True  (le maiuscole hanno codice minore delle minuscole)

print("---")

# Confronti concatenati — Python li supporta in modo elegante
x = 5
print(1 < x < 10)     # True  (equivale a 1 < x AND x < 10)
print(0 < x <= 5)     # True
print(1 < x < 4)      # False
````

## 3. Operatori logici

Combinano espressioni booleane. Sono tre: `and`, `or`, `not`.

### `and` — Entrambe le condizioni devono essere vere

```
True  and True   → True
True  and False  → False
False and True   → False
False and False  → False
```

### `or` — Almeno una condizione deve essere vera

```
True  or True   → True
True  or False  → True
False or True   → True
False or False  → False
```

### `not` — Inverte il valore

```
not True   → False
not False  → True
```

````python
# Tavole di verità complete
print("=== AND ===")
for a in [True, False]:
    for b in [True, False]:
        print(f"  {str(a):<6} and {str(b):<6} → {a and b}")

print("\n=== OR ===")
for a in [True, False]:
    for b in [True, False]:
        print(f"  {str(a):<6} or  {str(b):<6} → {a or b}")

print("\n=== NOT ===")
for a in [True, False]:
    print(f"  not {str(a):<6} → {not a}")
````

````python
# Casi pratici
età              = 20
ha_patente       = True
piove            = False
ha_ombrello      = True

# AND: entrambe le condizioni devono verificarsi
può_guidare = età >= 18 and ha_patente
print(f"Può guidare?          {può_guidare}")

# OR: basta che una si verifichi
esce_senza_bagnarsi = not piove or ha_ombrello
print(f"Esce senza bagnarsi?  {esce_senza_bagnarsi}")

# NOT: invertire una condizione
non_piove = not piove
print(f"Non piove?            {non_piove}")

# Combinazione complessa
può_andare_al_parco = età >= 18 and (not piove or ha_ombrello)
print(f"Può andare al parco?  {può_andare_al_parco}")
````

### 3.1 Valutazione a cortocircuito

Python è intelligente: in un'espressione `and` o `or`, smette di valutare non appena può determinare il risultato.

- `False and (...)` → il risultato è già `False`, il secondo operando **non viene valutato**.
- `True or (...)` → il risultato è già `True`, il secondo operando **non viene valutato**.

Questo è importante per evitare errori — ad esempio, verificare che un numero non sia zero **prima** di dividere:

````python
# Senza cortocircuito, la divisione per zero darebbe un errore
# Con and, se b == 0 la seconda parte non viene valutata
a = 10
b = 0

# Modo sicuro grazie al cortocircuito:
risultato = b != 0 and a / b > 2
print(f"b != 0 and a/b > 2  →  {risultato}")  # False (senza errore)

# Funziona anche con or:
nome = ""
nome_da_mostrare = nome or "Utente anonimo"   # se nome è vuoto (falsy), usa il valore predefinito
print(f"Nome da mostrare: {nome_da_mostrare}")
````

## 4. Operatori speciali

### 4.1 Operatore di identità: `is` e `is not`

`is` verifica se due variabili puntano allo **stesso oggetto in memoria** (non se hanno lo stesso valore).

```python
a == b    # hanno lo stesso valore?
a is b    # sono lo stesso oggetto in memoria?
```

L'uso più comune è verificare se qualcosa è `None`:
```python
risultato is None       # corretto
risultato == None       # funziona ma non è idiomatico
```

### 4.2 Operatore di appartenenza: `in` e `not in`

`in` verifica se un elemento è presente all'interno di una sequenza (string, lista, ecc.).

````python
# Operatore 'in' con le stringhe
frase = "Python è fantastico"
print("Python" in frase)       # True
print("Java" in frase)         # False
print("Java" not in frase)     # True

print("---")

# is None — modo corretto di verificare l'assenza di valore
valore = None
print(valore is None)          # True
print(valore is not None)      # False

print("---")

# Differenza tra == e is
a = [1, 2, 3]
b = [1, 2, 3]   # stessi valori, oggetto diverso
c = a           # stesso riferimento

print(f"a == b  →  {a == b}")   # True  (stesso contenuto)
print(f"a is b  →  {a is b}")   # False (oggetti distinti in memoria)
print(f"a is c  →  {a is c}")   # True  (c punta allo stesso oggetto di a)
````

## 5. Espressioni

Un'**espressione** è qualsiasi combinazione di valori, variabili e operatori che Python può **valutare** per produrre un risultato.

```python
2 + 3              # espressione aritmetica → 5
x > 0              # espressione di confronto → True o False
a and b or not c   # espressione logica → True o False
"ciao" + nome      # espressione con stringhe → una nuova stringa
```

Un'**istruzione** è un comando completo che Python esegue ma che non necessariamente produce un valore:
```python
x = 5              # istruzione di assegnazione
print(x)           # istruzione di chiamata a funzione
```

### Precedenza completa degli operatori

Quando mescoli operatori aritmetici, di confronto e logici:

| Priorità | Operatori |
|-----------|------------|
| 1 (massima) | `()` |
| 2 | `**` |
| 3 | `-x` (negazione) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `is`, `in` |
| 7 | `not` |
| 8 | `and` |
| 9 (minima) | `or` |

````python
# Python valuta: aritmetica → confronto → logica
x = 8

# Passo passo: prima 3*2=6, poi 8>6 → True, poi True and True → True
risultato = x > 3 * 2 and x < 100
print(f"x > 3*2 and x < 100  →  {risultato}")

# not ha precedenza minore dei confronti ma maggiore di and
print(not x == 8)       # False  (prima x==8 → True, poi not True → False)
print(not x == 8 or True)  # True

print("---")

# Espressione condizionale (ternaria) — forma compatta di if/else
# valore_se_vero  if  condizione  else  valore_se_falso
età = 20
stato = "maggiorenne" if età >= 18 else "minorenne"
print(f"Con {età} anni sei {stato}")

# Un altro esempio
numero = -5
segno = "positivo" if numero >= 0 else "negativo"
print(f"{numero} è {segno}")
````

### 5.1 L'operatore tricheco `:=` (Python 3.8+)

L'operatore `:=` (chiamato *walrus*, tricheco, per la sua somiglianza con la zanna dell'animale 🦭) permette di **assegnare e usare un valore nella stessa espressione**. È utile per evitare di calcolare qualcosa due volte.

````python
import math

# Senza walrus: calcoliamo e poi verifichiamo
raggio = 7
area = math.pi * raggio ** 2
if area > 100:
    print(f"Area grande: {area:.2f}")

# Con walrus: calcoliamo e verifichiamo in un solo passo
if (area := math.pi * raggio ** 2) > 100:
    print(f"Area grande (walrus): {area:.2f}")
````

n = int(input("Inserisci un numero intero: "))

- `div_per_3 and div_per_5` equivale a `n % 15 == 0`, ma è più esplicito.

Usa l'operatore ternario per mostrare la categoria in una sola riga.

Peso (kg): 70

peso    = float(input("Peso (kg): "))

**Punti chiave:**

- `not password.islower()` sfrutta il fatto che `.islower()` restituisce `True` solo se **tutti** i caratteri sono minuscoli — negandolo, rileviamo che c'è almeno una maiuscola.
- `è_valida = A and B and C` combina le tre condizioni — è `True` solo se tutte e tre sono `True`.

### Esercizio 5 (sfida) — Calcolatrice scientifica

**Consegna:** Scrivi una calcolatrice che risolva l'**equazione di secondo grado**:

```
ax² + bx + c = 0
```

Il programma deve:
1. Chiedere i coefficienti `a`, `b`, `c`.
2. Calcolare il **discriminante**: `D = b² - 4ac`
3. Usando operatori di confronto e logici, determinare:
   - Se `D > 0`: due soluzioni reali
   - Se `D == 0`: una soluzione (radice doppia)
   - Se `D < 0`: nessuna soluzione reale

```python
import math

a = float(input("Coefficiente a: "))
b = float(input("Coefficiente b: "))
c = float(input("Coefficiente c: "))

D = b**2 - 4*a*c

print(f"\nDiscriminante D = {D}")

ha_due_soluzioni     = D > 0
ha_una_soluzione     = D == 0
senza_soluzione_reale = D < 0

if ha_due_soluzioni:
    x1 = (-b + math.sqrt(D)) / (2*a)
    x2 = (-b - math.sqrt(D)) / (2*a)
    print("Due soluzioni reali:")
    print(f"  x1 = {x1}")
    print(f"  x2 = {x2}")
elif ha_una_soluzione:
    x = -b / (2*a)
    print(f"Una soluzione (radice doppia): x = {x}")
else:
    print("Nessuna soluzione reale (D < 0)")
```
