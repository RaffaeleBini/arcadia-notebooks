## 4. L'interprete di Python

### Come funziona Python?

Python è un linguaggio **interpretato**. Questo significa che non devi compilare il codice (trasformarlo in un eseguibile) prima di eseguirlo. Python legge ed esegue il tuo codice riga per riga in tempo reale.

```
Il tuo codice (.py o cella)  →  Interprete di Python  →  Risultato
     print("Ciao")           →       legge ed esegue   →    Ciao
```

### Confronto rapido

| | Python (interpretato) | C++ (compilato) |
|--|----------------------|------------------|
| Velocità di sviluppo | Rapida | Lenta |
| Velocità di esecuzione | Più lenta | Più rapida |
| Facilità di apprendimento | Alta | Bassa |
| Uso in IA/Data Science | ⭐⭐⭐⭐⭐ | ⭐⭐ |

### Python nel terminale (modalità interattiva)

Puoi anche usare Python direttamente nel terminale scrivendo `python` e premendo Invio. Vedrai il prompt `>>>` — puoi scrivere codice riga per riga.

```
$ python
>>> 2 + 2
4
>>> print("Ciao")
Ciao
>>> exit()
```

Nei notebook, ogni cella di codice funziona esattamente come questa modalità interattiva.

## 5. Il tuo primo programma

### 5.1 `print()` — Mostrare informazioni a schermo

`print()` è la funzione più basilare di Python. Mostra testo o valori nell'output.

```python
print("Ciao, mondo")   # mostra testo
print(42)              # mostra un numero
print(3.14)            # mostra un decimale
```

Le parole tra virgolette (`"..."` o `'...'`) si chiamano **stringhe di testo** o **string**.

````python
print("Ciao, mondo")
print("Il mio nome è Python")
print("L'anno è:", 2025)
````

### 5.2 Variabili — Salvare informazioni

Una **variabile** è come una scatola con un'etichetta dove salvi un valore per usarlo in seguito.

```
  nome = "Ana"
  ┌───────┐        ┌─────────┐
  │  nome │  ────▶ │  "Ana"  │
  └───────┘        └─────────┘
   etichetta          valore
```

**Regole per nominare le variabili:**
- Solo lettere, numeri e trattini bassi `_`
- Non possono iniziare con un numero
- Niente spazi (usa `mia_variabile` invece di `mia variabile`)
- Distingue maiuscole e minuscole: `nome` e `Nome` sono variabili diverse

**Convenzione:** usa `snake_case` (parole separate da `_`) per i nomi.

````python
# Creiamo delle variabili
nome = "Ana"
eta = 28
altezza = 1.65
e_studentessa = True

# Le usiamo
print(nome)
print(eta)
print(altezza)
print(e_studentessa)
````

### 5.3 Tipi di dati di base

Ogni valore in Python ha un **tipo**. I più fondamentali sono:

| Tipo | Nome in Python | Esempio | A cosa serve |
|------|-----------------|---------|------------------|
| Numero intero | `int` | `42`, `-7`, `0` | Contare, indicizzare |
| Numero decimale | `float` | `3.14`, `-0.5` | Misure, calcoli |
| Testo | `str` | `"ciao"`, `'mondo'` | Messaggi, nomi |
| Vero/Falso | `bool` | `True`, `False` | Condizioni logiche |

Puoi conoscere il tipo di qualsiasi valore con la funzione `type()`.

````python
# Verifichiamo il tipo di ogni variabile
print(type(42))          # int
print(type(3.14))        # float
print(type("ciao"))      # str
print(type(True))        # bool

# Funziona anche con le variabili
eta = 28
print(type(eta))
````

### 5.4 Operatori aritmetici

Python può eseguire qualsiasi operazione matematica:

| Operatore | Simbolo | Esempio | Risultato |
|----------|---------|---------|----------|
| Somma | `+` | `3 + 2` | `5` |
| Sottrazione | `-` | `10 - 4` | `6` |
| Moltiplicazione | `*` | `3 * 4` | `12` |
| Divisione | `/` | `10 / 3` | `3.333...` |
| Divisione intera | `//` | `10 // 3` | `3` |
| Modulo (resto) | `%` | `10 % 3` | `1` |
| Potenza | `**` | `2 ** 8` | `256` |

````python
a = 10
b = 3

print("Somma:              ", a + b)
print("Sottrazione:        ", a - b)
print("Moltiplicazione:    ", a * b)
print("Divisione:          ", a / b)
print("Divisione intera:   ", a // b)
print("Modulo (resto):     ", a % b)
print("Potenza:            ", a ** b)
````

### 5.5 `input()` — Ricevere dati dall'utente

`input()` ferma il programma e aspetta che l'utente scriva qualcosa. Quello che scrive viene salvato come **string**.

```python
nome = input("Come ti chiami? ")
print("Ciao,", nome)
```

> ⚠️ **Importante:** `input()` restituisce sempre testo (`str`). Se hai bisogno di un numero, devi convertirlo con `int()` o `float()`.

```python
eta = int(input("Quanti anni hai? "))
prezzo = float(input("Quanto costa? "))
```

````python
# Programma che saluta l'utente
nome = input("Come ti chiami? ")
print("Ciao,", nome + "! Benvenuto/a al corso di IA.")
````

````python
# Conversione di tipi
anno_nascita = int(input("In che anno sei nato/a? "))
eta_approssimativa = 2025 - anno_nascita
print("Hai circa", eta_approssimativa, "anni.")
````

### 5.6 f-string — Testo con variabili integrate

Una **f-string** è il modo più comodo per combinare testo e variabili. Si scrive mettendo una `f` prima delle virgolette, e le variabili vanno tra `{}`.

```python
nome = "Ana"
eta = 28
print(f"Mi chiamo {nome} e ho {eta} anni.")
# → Mi chiamo Ana e ho 28 anni.
```

Puoi anche inserire espressioni dentro `{}`:
```python
print(f"Il doppio di 5 è {5 * 2}")
```

````python
nome = "Ana"
eta = 28
citta = "Madrid"

print(f"Mi chiamo {nome}, ho {eta} anni e vivo a {citta}.")
print(f"L'anno prossimo avrò {eta + 1} anni.")
print(f"Il doppio della mia età è {eta * 2}.")
````

### Esercizio 1 — Calcolatrice di due numeri

**Consegna:** Scrivi un programma che:
1. Chieda all'utente due numeri.
2. Mostri la somma, la sottrazione, la moltiplicazione e la divisione di quei numeri.

**Esempio di output atteso:**
```
Inserisci il primo numero: 10
Inserisci il secondo numero: 4

--- Risultati ---
10.0 + 4.0 = 14.0
10.0 - 4.0 = 6.0
10.0 * 4.0 = 40.0
10.0 / 4.0 = 2.5
```

````python
# ✏️ IL TUO CODICE QUI
````

<details>
<summary>💡 Vedi soluzione — Esercizio 1</summary>

```python
# Chiediamo i due numeri all'utente
num1 = float(input("Inserisci il primo numero: "))
num2 = float(input("Inserisci il secondo numero: "))

# Calcoliamo le operazioni
somma          = num1 + num2
sottrazione    = num1 - num2
prodotto       = num1 * num2
divisione      = num1 / num2

# Mostriamo i risultati
print("\n--- Risultati ---")
print(f"{num1} + {num2} = {somma}")
print(f"{num1} - {num2} = {sottrazione}")
print(f"{num1} * {num2} = {prodotto}")
print(f"{num1} / {num2} = {divisione}")
```

**Punti chiave:**
- Usiamo `float()` per permettere all'utente di inserire decimali.
- `\n` dentro una string stampa una riga vuota.
- Le f-string ci permettono di mostrare l'operazione e il risultato sulla stessa riga in modo leggibile.
</details>

**Consegna:** Modifica la calcolatrice affinché mostri anche:
- La **divisione intera** (quante volte il secondo numero sta nel primo).
- Il **resto** di quella divisione.

**Esempio di output atteso:**
```
Inserisci il primo numero: 17
```

```python
num1 = int(input("Inserisci il primo numero: "))
```

### Esercizio 3 — Calcolatrice di area e perimetro

**Consegna:** Scrivi un programma che chieda il **raggio di un cerchio** e calcoli:
- La sua **area** (formula: π × r²)
- Il suo **perimetro** (formula: 2 × π × r)

Usa `3.14159` come valore di π (oppure importa `math.pi` se vuoi essere più preciso).

**Esempio di output atteso:**
```
Inserisci il raggio del cerchio: 5

Cerchio con raggio 5.0:
  Area:      78.53975
  Perimetro: 31.4159
```

````python
# ✏️ IL TUO CODICE QUI
# Suggerimento: per usare pi greco dalla libreria math, scrivi all'inizio:
# import math
# e poi usa math.pi invece di 3.14159
````

<details>
<summary>💡 Vedi soluzione — Esercizio 3</summary>

```python
import math

raggio = float(input("Inserisci il raggio del cerchio: "))

area      = math.pi * raggio ** 2
perimetro = 2 * math.pi * raggio

print(f"\nCerchio con raggio {raggio}:")
print(f"  Area:      {area:.5f}")
print(f"  Perimetro: {perimetro:.5f}")
```

**Punti chiave:**
- `import math` carica la libreria matematica di Python. Le librerie sono raccolte di funzioni già scritte che puoi riutilizzare.
- `math.pi` è il valore di π con la massima precisione.
- `{area:.5f}` dentro una f-string formatta il numero con 5 decimali. Il formato è `{valore:.Nf}` dove N è il numero di decimali.
</details>

### Esercizio 4 (sfida) — Calcolatrice di mance

**Consegna:** In un ristorante, scrivi un programma che:
1. Chieda il **totale del conto** in euro.
2. Chieda la **percentuale di mancia** che l'utente vuole lasciare (es: 10 per 10%).
3. Mostri:
   - L'importo della mancia.
   - Il totale da pagare (conto + mancia).
   - Se ci sono 4 persone, **quanto paga ciascuna?**

**Esempio di output atteso:**
```
Totale del conto (€): 85.50
Percentuale di mancia (%): 15
Numero di persone: 4

--- Riepilogo ---
Conto:           85.50 €
Mancia (15%):    12.82 €
Totale:          98.32 €
A persona:       24.58 €
```

````python
# ✏️ IL TUO CODICE QUI
````

<details>
<summary>💡 Vedi soluzione — Esercizio 4</summary>

```python
conto       = float(input("Totale del conto (€): "))
percentuale = float(input("Percentuale di mancia (%): "))
persone     = int(input("Numero di persone: "))

mancia      = conto * (percentuale / 100)
totale      = conto + mancia
a_persona   = totale / persone

print("\n--- Riepilogo ---")
print(f"Conto:           {conto:.2f} €")
print(f"Mancia ({percentuale:.0f}%):    {mancia:.2f} €")
print(f"Totale:          {totale:.2f} €")
print(f"A persona:       {a_persona:.2f} €")
```

**Punti chiave:**
- `percentuale / 100` converte la percentuale in un fattore moltiplicativo (15% → 0.15).
- `{valore:.2f}` formatta con esattamente 2 decimali, ideale per le valute.
- `{percentuale:.0f}` mostra la percentuale senza decimali anche se è `float`.
</details>
