## 1. El problema: el código necesita decidir

Imagina que quieres escribir un programa que diga si una persona puede votar. Sin condicionales, no puedes — el código haría lo mismo para cualquier edad:

```python
edad = 15
print("Puedes votar")   # esto se imprime siempre, aunque la edad sea 15
```

Necesitamos una estructura que diga: **«si la edad es mayor o igual a 18, entonces muestra el mensaje»**.

Eso es exactamente lo que hace `if`.

---
## 2. `if` — Si se cumple la condición, ejecuta este bloque

### Sintaxis

```python
if condicion:
    # bloque de código
    # estas líneas solo se ejecutan si condicion es True
```

**Reglas obligatorias:**
1. La condición va seguida de **dos puntos** `:`
2. El bloque interior va **indentado** — desplazado 4 espacios (o 1 tabulación) respecto al `if`
3. Python usa la **indentación** para saber qué código pertenece al bloque — no usa llaves `{}` como otros lenguajes

```
if condicion:          ← dos puntos obligatorios
    instruccion_1      ← indentada (pertenece al if)
    instruccion_2      ← indentada (pertenece al if)
instruccion_3          ← NO indentada (se ejecuta siempre)
```

````python
edad = 20

if edad >= 18:
    print("Puedes votar")           # solo se ejecuta si edad >= 18
    print("Recuerda llevar el DNI") # también pertenece al if

print("Programa terminado")         # esto se ejecuta siempre
````

````python
# Prueba cambiando el valor de edad a 15 y vuelve a ejecutar
edad = 15

if edad >= 18:
    print("Puedes votar")
    print("Recuerda llevar el DNI")

print("Programa terminado")   # esto sí aparece aunque edad sea 15
````

### ⚠️ La indentación es obligatoria y precisa

La indentación incorrecta es uno de los errores más comunes en Python. Veamos los tipos de error:

````python
# Error 1: falta indentación → IndentationError
# Descomenta para ver el error:
# if True:
# print("falta indentación")   ← IndentationError

# Error 2: indentación inconsistente
# if True:
#     print("4 espacios")
#       print("6 espacios")    ← IndentationError

# CORRECTO: 4 espacios consistentes (o usar Tab)
x = 10
if x > 5:
    print("x es mayor que 5")
    print("esto también está en el bloque")
print("esto está fuera del bloque")
````

## 3. `else` — Qué hacer cuando la condición NO se cumple

`else` define un bloque que se ejecuta **solo cuando la condición del `if` es `False`**. Siempre va emparejado con un `if`.

```python
if condicion:
    # se ejecuta si condicion es True
else:
    # se ejecuta si condicion es False
```

Solo uno de los dos bloques se ejecuta — nunca los dos a la vez.

````python
edad = 16

if edad >= 18:
    print("Acceso permitido")
else:
    print("Acceso denegado: debes tener al menos 18 años")

print("---")

# Otro ejemplo: número positivo o negativo
numero = -7

if numero >= 0:
    print(f"{numero} es positivo (o cero)")
else:
    print(f"{numero} es negativo")
````

````python
# Ejemplo práctico: división segura
a = 10
b = 0

if b != 0:
    resultado = a / b
    print(f"{a} / {b} = {resultado}")
else:
    print("Error: no se puede dividir entre cero")
````

## 4. `elif` — Múltiples caminos

Cuando hay más de dos posibilidades, usamos `elif` (abreviatura de *else if*). Python evalúa las condiciones **de arriba hacia abajo** y ejecuta solo el primer bloque cuya condición sea `True`. El resto se salta.

```python
if condicion_1:
    # bloque 1
elif condicion_2:
    # bloque 2
elif condicion_3:
    # bloque 3
else:
    # bloque por defecto (ninguna condición anterior fue True)
```

**Puntos clave:**
- Puede haber **cero o más** `elif` entre el `if` y el `else`.
- El `else` al final es **opcional** — pero si lo pones, actúa como red de seguridad.
- Solo se ejecuta **uno** de todos los bloques.

````python
# Clasificación por nota
nota = 7.5

if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(f"Nota: {nota} → {calificacion}")
````

````python
# ⚠️ El ORDEN de los elif importa
# Python para en la primera condición verdadera

nota = 9.5

# ORDEN CORRECTO: de mayor a menor
if nota >= 9:
    print(f"Correcto → Sobresaliente ({nota})")
elif nota >= 7:
    print(f"Notable ({nota})")
elif nota >= 5:
    print(f"Aprobado ({nota})")

print("---")

# ORDEN INCORRECTO: de menor a mayor → error lógico silencioso
if nota >= 5:
    print(f"Incorrecto → Aprobado ({nota})")   # se ejecuta aunque sea 9.5
elif nota >= 7:
    print(f"Notable ({nota})")                 # nunca llega aquí
elif nota >= 9:
    print(f"Sobresaliente ({nota})")           # nunca llega aquí
````

````python
# Ejemplo: clasificación de temperatura
temperatura = float(input("¿Cuál es la temperatura exterior (°C)? "))

if temperatura <= 0:
    descripcion = "helada 🧊"
    consejo     = "Abrígate mucho"
elif temperatura <= 10:
    descripcion = "muy fría"
    consejo     = "Lleva abrigo"
elif temperatura <= 20:
    descripcion = "fresca"
    consejo     = "Una chaqueta puede ir bien"
elif temperatura <= 28:
    descripcion = "agradable"
    consejo     = "Temperatura ideal"
else:
    descripcion = "calurosa 🌡️"
    consejo     = "Hidrátate bien"

print(f"\nTemperatura {descripcion} ({temperatura}°C)")
print(f"Consejo: {consejo}")
````

## 5. Anidamiento — `if` dentro de `if`

Puedes poner un `if` dentro del bloque de otro `if`. Esto se llama **anidamiento** y sirve para comprobar condiciones más específicas.

```python
if condicion_exterior:
    # solo llegamos aquí si condicion_exterior es True
    if condicion_interior:
        # solo llegamos aquí si las dos son True
    else:
        # condicion_exterior True, condicion_interior False
else:
    # condicion_exterior es False
```

> ⚠️ **Precaución:** no anides más de 2-3 niveles. El código se vuelve difícil de leer. Si tienes mucho anidamiento, probablemente puedes reescribir las condiciones con `and`.

````python
# Ejemplo: acceso a un sistema
usuario_activo = True
contrasena_ok  = True
es_admin       = False

if usuario_activo:
    if contrasena_ok:
        print("Acceso concedido")
        if es_admin:
            print("Bienvenido al panel de administración")
        else:
            print("Bienvenido al área de usuario")
    else:
        print("Contraseña incorrecta")
else:
    print("Cuenta desactivada")
````

````python
# La misma lógica, escrita de forma más plana con 'and'
# Más legible cuando las condiciones son independientes

usuario_activo = True
contrasena_ok  = True
es_admin       = False

if not usuario_activo:
    print("Cuenta desactivada")
elif not contrasena_ok:
    print("Contraseña incorrecta")
elif es_admin:
    print("Acceso concedido — panel de administración")
else:
    print("Acceso concedido — área de usuario")
````

## 6. Buenas prácticas

### 6.1 Condiciones legibles

Evita condiciones innecesariamente complicadas:

````python
activo = True

# ❌ Redundante — comparar un bool con True es innecesario
if activo == True:
    print("activo (forma redundante)")

# ✅ Idiomático — el bool ya es la condición
if activo:
    print("activo (forma correcta)")

# ❌ Redundante para False
if activo == False:
    print("no activo (redundante)")

# ✅ Usa 'not'
if not activo:
    print("no activo (correcto)")
````

````python
# 6.2 Clausulas de guarda (guard clauses)
# En lugar de anidar, descarta los casos inválidos al principio

# ❌ Lógica muy anidada
def procesar_v1(numero):
    if numero is not None:
        if numero > 0:
            if numero % 2 == 0:
                print(f"{numero} es un entero positivo par")

# ✅ Clausulas de guarda: descarta casos inválidos antes
def procesar_v2(numero):
    if numero is None:
        print("El número no puede ser None")
        return
    if numero <= 0:
        print("El número debe ser positivo")
        return
    if numero % 2 != 0:
        print("El número debe ser par")
        return
    print(f"{numero} es un entero positivo par")

procesar_v1(8)
procesar_v2(8)
procesar_v2(-3)
procesar_v2(7)
````

````python
# 6.3 match/case — Python 3.10+
# Una alternativa más limpia al if/elif cuando comparas una variable
# contra valores concretos (como un switch de otros lenguajes)

dia = "lunes"

# Con if/elif
if dia == "lunes" or dia == "martes" or dia == "miércoles" or dia == "jueves" or dia == "viernes":
    tipo = "laborable"
elif dia == "sábado" or dia == "domingo":
    tipo = "fin de semana"
else:
    tipo = "día desconocido"
print(f"{dia}: {tipo}")

# Con match/case (más elegante)
match dia:
    case "lunes" | "martes" | "miércoles" | "jueves" | "viernes":
        tipo = "laborable"
    case "sábado" | "domingo":
        tipo = "fin de semana"
    case _:            # _ es el caso por defecto (como else)
        tipo = "día desconocido"

print(f"{dia}: {tipo}")
````

mate    = float(input("Nota de Matemáticas: "))

media            = float(input("Nota media: "))

- El `else` actúa como red de seguridad: captura cualquier caso que no encaje en las becas anteriores.
- Separar el cálculo (`if/elif/else`) de la presentación (`print`) hace el código más organizado y fácil de modificar.

## Resumen de la clase

### Estructura general

```python
if condicion_1:           # obligatorio, siempre primero
    bloque_1
elif condicion_2:         # opcional, puede repetirse
    bloque_2
elif condicion_3:
    bloque_3
else:                     # opcional, siempre al final
    bloque_por_defecto
```

### Reglas esenciales

| Regla | Detalle |
|-------|---------|
| Dos puntos `:` | Obligatorios tras `if`, `elif` y `else` |
| Indentación | 4 espacios — define qué pertenece al bloque |
| Solo un bloque se ejecuta | Python para en la primera condición verdadera |
| Orden de condiciones | De la más restrictiva a la más general |
| `else` | Opcional — captura el caso por defecto |

### Patrones habituales

```python
# Validación primero
if entrada_invalida:
    print("Error")
else:
    # lógica principal

# Booleano directo (sin comparar con True/False)
if activo:          # no: if activo == True
if not activo:      # no: if activo == False

# Ternario para valores simples
etiqueta = "par" if n % 2 == 0 else "impar"
```
