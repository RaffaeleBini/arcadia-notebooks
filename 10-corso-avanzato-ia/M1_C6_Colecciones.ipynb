## 1. Listas (`list`)

Una lista es una secuencia **ordenada** y **mutable** de elementos. Es la colección más usada en Python.

- **Ordenada:** los elementos tienen una posición fija (índice).
- **Mutable:** puedes añadir, eliminar o cambiar elementos después de crearla.
- **Permite duplicados:** el mismo valor puede aparecer varias veces.
- **Heterogénea:** puede contener elementos de distintos tipos.

```python
notas       = [8.5, 6.0, 9.0, 5.5]         # lista de floats
nombres     = ["Ana", "Luis", "María"]      # lista de strings
mixta       = [1, "hola", True, 3.14]       # tipos mezclados
vacia       = []                             # lista vacía
anidada     = [[1, 2], [3, 4], [5, 6]]      # lista de listas
```

````python
# Crear listas
notas   = [8.5, 6.0, 9.0, 5.5, 7.0]
nombres = ["Ana", "Luis", "María", "Carlos"]
vacia   = []

print(notas)
print(nombres)
print(type(notas))       # <class 'list'>
print(len(notas))        # número de elementos
````

### 1.1 Acceso por índice y slicing

Igual que los strings, las listas usan índices desde `0`. Los índices negativos cuentan desde el final.

```
lista  =  ["a",  "b",  "c",  "d",  "e"]
índice:     0     1     2     3     4
negativo:  -5    -4    -3    -2    -1
```

El **slicing** `lista[inicio:fin:paso]` devuelve una sublista (el índice `fin` no se incluye).

````python
frutas = ["manzana", "pera", "uva", "naranja", "kiwi"]

# Acceso por índice
print(frutas[0])    # manzana
print(frutas[-1])   # kiwi (último)
print(frutas[2])    # uva

# Slicing
print(frutas[1:3])  # ['pera', 'uva']
print(frutas[:2])   # ['manzana', 'pera']
print(frutas[3:])   # ['naranja', 'kiwi']
print(frutas[::-1]) # lista al revés
````

### 1.2 Modificar listas

Las listas son **mutables**: puedes cambiar, añadir y eliminar elementos.

| Operación | Método / Sintaxis | ¿Qué hace? |
|-----------|------------------|------------|
| Cambiar elemento | `lista[i] = valor` | Sustituye el elemento en posición `i` |
| Añadir al final | `lista.append(x)` | Añade `x` al final |
| Insertar en posición | `lista.insert(i, x)` | Inserta `x` en la posición `i` |
| Añadir otra lista | `lista.extend(otra)` | Añade todos los elementos de `otra` |
| Eliminar por valor | `lista.remove(x)` | Elimina la primera aparición de `x` |
| Eliminar por índice | `lista.pop(i)` | Elimina y devuelve el elemento en `i` |
| Vaciar | `lista.clear()` | Elimina todos los elementos |
| Eliminar (sin retorno) | `del lista[i]` | Elimina el elemento en `i` |

````python
colores = ["rojo", "verde", "azul"]
print(f"Inicial:          {colores}")

colores.append("amarillo")
print(f"Tras append:      {colores}")

colores.insert(1, "naranja")
print(f"Tras insert(1):   {colores}")

colores.remove("verde")
print(f"Tras remove:      {colores}")

eliminado = colores.pop(0)
print(f"Tras pop(0):      {colores}  ← eliminado: '{eliminado}'")

colores[0] = "morado"
print(f"Tras cambiar [0]: {colores}")
````

### 1.3 Otros métodos y operaciones útiles

````python
numeros = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]

print(f"Lista original:   {numeros}")
print(f"len():            {len(numeros)}")
print(f"sum():            {sum(numeros)}")
print(f"max():            {max(numeros)}")
print(f"min():            {min(numeros)}")
print(f"count(5):         {numeros.count(5)}")   # cuántas veces aparece 5
print(f"index(9):         {numeros.index(9)}")   # posición de la primera aparición
print(f"'4' in lista:     {4 in numeros}")

# Ordenar
copia = numeros.copy()   # copia para no modificar el original
copia.sort()
print(f"sorted (asc):     {copia}")
copia.sort(reverse=True)
print(f"sorted (desc):    {copia}")

# sorted() devuelve una nueva lista sin modificar la original
print(f"sorted() nueva:   {sorted(numeros)}")
print(f"Original intacta: {numeros}")
````

### 1.4 List comprehension — Crear listas de forma compacta

Una **list comprehension** es una forma elegante de crear una lista aplicando una expresión a cada elemento de otra secuencia. Es uno de los patrones más idiomáticos de Python.

```python
# Forma larga:
cuadrados = []
for n in range(1, 6):
    cuadrados.append(n ** 2)

# Equivalente con comprehension:
cuadrados = [n ** 2 for n in range(1, 6)]
```

Sintaxis: `[expresión  for  variable  in  secuencia  if  condición]`

````python
# Cuadrados del 1 al 10
cuadrados = [n ** 2 for n in range(1, 11)]
print(f"Cuadrados:     {cuadrados}")

# Solo los pares
pares = [n for n in range(1, 21) if n % 2 == 0]
print(f"Pares 1-20:    {pares}")

# Transformar strings
nombres   = ["ana", "luis", "maría", "carlos"]
en_mayus  = [n.capitalize() for n in nombres]
print(f"Capitalizados: {en_mayus}")

# Filtrar notas aprobadas
notas     = [4, 8, 3, 9, 5, 2, 7, 6]
aprobadas = [n for n in notas if n >= 5]
print(f"Aprobadas:     {aprobadas}")
````

## 2. Tuplas (`tuple`)

Una tupla es como una lista, pero **inmutable**: una vez creada, no puedes cambiarla. Se escribe con paréntesis.

```python
coordenadas = (40.4168, -3.7038)   # latitud, longitud de Madrid
colores_rgb = (255, 128, 0)        # rojo, verde, azul
vacia       = ()
un_elemento = (42,)                # ← coma obligatoria para tuplas de 1 elemento
```

### ¿Cuándo usar tupla en lugar de lista?

| Situación | Usa |
|-----------|-----|
| Los datos **no deben cambiar** (coordenadas, configuración, constantes) | `tuple` |
| Los datos **pueden cambiar** (colección dinámica) | `list` |
| Como **clave de diccionario** (las claves deben ser inmutables) | `tuple` |
| Devolver **múltiples valores** desde una función | `tuple` |

````python
# Crear tuplas
coordenadas = (40.4168, -3.7038)
rgb_rojo    = (255, 0, 0)

print(coordenadas)
print(type(coordenadas))    # <class 'tuple'>

# Acceso igual que las listas
print(coordenadas[0])       # latitud
print(coordenadas[1])       # longitud

# Desempaquetado (unpacking): asignar cada elemento a una variable
latitud, longitud = coordenadas
print(f"Latitud: {latitud}, Longitud: {longitud}")

# Intercambiar dos variables con tupla (patrón elegante de Python)
a, b = 10, 20
a, b = b, a    # swap sin variable temporal
print(f"a={a}, b={b}")
````

````python
# Las tuplas son inmutables — intentar modificarlas da error
punto = (3, 7)

# Esto funciona:
print(punto[0])
print(len(punto))
print(3 in punto)

# Descomenta para ver el error:
# punto[0] = 10   # TypeError: 'tuple' object does not support item assignment
````

````python
# Desempaquetado avanzado con * (captura el resto)
primero, *resto = (1, 2, 3, 4, 5)
print(f"Primero: {primero}")   # 1
print(f"Resto:   {resto}")     # [2, 3, 4, 5]

*inicio, ultimo = (1, 2, 3, 4, 5)
print(f"Inicio:  {inicio}")    # [1, 2, 3, 4]
print(f"Último:  {ultimo}")    # 5

# Uso típico: función que devuelve varios valores
def minmax(lista):
    return min(lista), max(lista)  # devuelve una tupla

minimo, maximo = minmax([4, 7, 2, 9, 1])
print(f"Mínimo: {minimo}, Máximo: {maximo}")
````

## 3. Conjuntos (`set`)

Un conjunto es una colección **desordenada** y **sin elementos duplicados**. Se escribe con llaves (pero sin claves).

```python
colores = {"rojo", "verde", "azul"}
```

Sus dos superpoderes:
1. **Elimina duplicados automáticamente**
2. **Operaciones de teoría de conjuntos** (unión, intersección, diferencia)

> ⚠️ `{}` crea un diccionario vacío, no un set. Para un set vacío usa `set()`.

````python
# Crear sets
colores = {"rojo", "verde", "azul", "rojo"}  # el duplicado desaparece
print(colores)          # el orden puede variar
print(type(colores))

# Convertir lista a set para eliminar duplicados
lista_con_dupes = [1, 2, 2, 3, 3, 3, 4]
sin_dupes = set(lista_con_dupes)
print(sin_dupes)

# Convertir de vuelta a lista
lista_limpia = sorted(sin_dupes)  # sorted() para ordenarlo
print(lista_limpia)

# Comprobar pertenencia — muy eficiente en sets
print("rojo" in colores)    # True
print("negro" in colores)   # False
````

````python
# Operaciones de conjuntos
matematicas = {"Ana", "Luis", "María", "Carlos"}
fisica      = {"Luis", "María", "Pedro", "Laura"}

# Unión: todos los que están en alguno de los dos
print(f"Unión (|):          {matematicas | fisica}")

# Intersección: solo los que están en los dos
print(f"Intersección (&):   {matematicas & fisica}")

# Diferencia: los que están en el primero pero no en el segundo
print(f"Diferencia (-):     {matematicas - fisica}")

# Diferencia simétrica: los que están en uno u otro, pero no en ambos
print(f"Dif. simétrica (^): {matematicas ^ fisica}")
````

````python
# Métodos de modificación
tecnologias = {"Python", "SQL", "Git"}

tecnologias.add("Docker")          # añade un elemento
print(f"Tras add:     {tecnologias}")

tecnologias.update({"AWS", "Linux"})  # añade varios a la vez
print(f"Tras update:  {tecnologias}")

tecnologias.discard("Git")         # elimina si existe (sin error si no existe)
print(f"Tras discard: {tecnologias}")

# issubset / issuperset
basicas  = {"Python", "SQL"}
print(f"¿basicas ⊆ tecnologias? {basicas.issubset(tecnologias)}")
````

## 4. Diccionarios (`dict`)

Un diccionario almacena pares **clave → valor**. En lugar de acceder por posición numérica, accedes por **clave** — cualquier valor inmutable (string, número, tupla).

```python
persona = {
    "nombre":  "Ana",
    "edad":    28,
    "ciudad":  "Madrid"
}
```

```
clave    →   valor
"nombre" →   "Ana"
"edad"   →   28
"ciudad" →   "Madrid"
```

- Las **claves** son únicas — no puede haber dos iguales.
- Los **valores** pueden ser de cualquier tipo, incluso otras listas o diccionarios.
- Desde Python 3.7, los diccionarios **mantienen el orden de inserción**.

````python
# Crear diccionario
persona = {
    "nombre": "Ana",
    "edad":   28,
    "ciudad": "Madrid"
}

print(persona)
print(type(persona))

# Acceso por clave
print(persona["nombre"])    # Ana
print(persona["edad"])      # 28

# .get() — acceso seguro: devuelve None (o un default) si la clave no existe
print(persona.get("email"))              # None
print(persona.get("email", "sin email")) # sin email
````

````python
# Modificar diccionarios
persona = {"nombre": "Ana", "edad": 28}

# Añadir o modificar
persona["ciudad"] = "Madrid"    # añade clave nueva
persona["edad"]   = 29          # modifica clave existente
print(f"Tras modificar: {persona}")

# Eliminar
del persona["ciudad"]
print(f"Tras del:       {persona}")

valor = persona.pop("edad")     # elimina y devuelve el valor
print(f"Tras pop:       {persona}  ← edad eliminada: {valor}")

# Fusionar dos diccionarios (Python 3.9+: operador |)
base  = {"nombre": "Ana", "edad": 28}
extra = {"ciudad": "Madrid", "email": "ana@ejemplo.com"}
completo = base | extra
print(f"Fusionado:      {completo}")
````

````python
# Iterar sobre diccionarios
persona = {
    "nombre": "Ana",
    "edad":   28,
    "ciudad": "Madrid",
    "email":  "ana@ejemplo.com"
}

# Solo claves
print("Claves:")
for clave in persona:
    print(f"  {clave}")

# Solo valores
print("Valores:")
for valor in persona.values():
    print(f"  {valor}")

# Claves y valores a la vez (lo más habitual)
print("Pares clave-valor:")
for clave, valor in persona.items():
    print(f"  {clave:10} → {valor}")
````

````python
# Diccionarios anidados — muy habitual para representar datos estructurados
agenda = {
    "ana": {
        "telefono": "600111222",
        "email":    "ana@ejemplo.com",
        "ciudad":   "Madrid"
    },
    "luis": {
        "telefono": "600333444",
        "email":    "luis@ejemplo.com",
        "ciudad":   "Barcelona"
    }
}

# Acceso anidado
print(agenda["ana"]["telefono"])   # 600111222

# Iterar
for nombre, datos in agenda.items():
    print(f"{nombre.capitalize()}: {datos['telefono']} ({datos['ciudad']})")
````

### 4.1 Dict comprehension

````python
# Dict comprehension: {clave: valor for ...}

# Cuadrados del 1 al 5
cuadrados = {n: n**2 for n in range(1, 6)}
print(cuadrados)    # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Invertir un diccionario (valor → clave)
original  = {"a": 1, "b": 2, "c": 3}
invertido = {v: k for k, v in original.items()}
print(invertido)    # {1: 'a', 2: 'b', 3: 'c'}

# Filtrar solo las notas aprobadas
notas = {"Ana": 8.5, "Luis": 4.0, "María": 7.0, "Carlos": 3.5}
aprobados = {alumno: nota for alumno, nota in notas.items() if nota >= 5}
print(aprobados)
````

## Comparativa de colecciones

| | `list` | `tuple` | `set` | `dict` |
|-|--------|---------|-------|--------|
| Se crea con | `[...]` | `(...)` | `{...}` | `{k: v}` |
| Ordenada | ✅ | ✅ | ❌ | ✅ (3.7+) |
| Mutable | ✅ | ❌ | ✅ | ✅ |
| Duplicados | ✅ | ✅ | ❌ | ❌ (claves) |
| Acceso por | índice | índice | — | clave |
| Caso de uso principal | Secuencias dinámicas | Datos fijos | Unicidad y teoría de conjuntos | Datos clave-valor |

```python
def añadir_contacto(agenda, nombre, telefono, email):
    clave = nombre.split()[0].lower()   # primera palabra del nombre como clave
    if clave in agenda:
        print(f"Ya existe un contacto con clave '{clave}'.")
        return
    agenda[clave] = {"nombre": nombre, "telefono": telefono, "email": email}
    print(f"Contacto '{nombre}' añadido.")

def editar_contacto(agenda, clave, campo, nuevo_valor):
    clave = clave.lower()
    if clave not in agenda:
        print(f"No existe ningún contacto con clave '{clave}'.")
        return
    campos_validos = {"nombre", "telefono", "email"}
    if campo not in campos_validos:
        print(f"Campo no válido. Campos disponibles: {campos_validos}")
        return
    agenda[clave][campo] = nuevo_valor
    print(f"Campo '{campo}' actualizado para {agenda[clave]['nombre']}.")

def eliminar_contacto(agenda, clave):
    clave = clave.lower()
    if clave not in agenda:
        print(f"No existe ningún contacto con clave '{clave}'.")
        return
    nombre = agenda[clave]['nombre']
    confirmar = input(f"¿Eliminar a {nombre}? (s/n): ")
    if confirmar.lower() == "s":
        del agenda[clave]
        print(f"Contacto '{nombre}' eliminado.")
    else:
        print("Eliminación cancelada.")

# Prueba

editar_contacto(agenda, "carlos", "telefono", "611999000")

```

**Puntos clave:**
- `campos_validos = {"nombre", "telefono", "email"}` es un **set** — lo usamos para validar que el campo existe. `in` en un set es más eficiente que en una lista.
- `clave not in agenda` comprueba la existencia antes de operar — evita `KeyError`.
- Separar la lógica en funciones permite llamar a `mostrar_agenda` en cualquier punto sin duplicar código.

```python
def buscar_por_etiqueta(agenda, etiqueta):
    etiqueta = etiqueta.lower()
```

```
Historial de Ana García:
  2025-01-10  →  120 s  (2 min 0 s)
```

```python
def registrar_llamada(agenda, clave, fecha, duracion):
    clave = clave.lower()
    if clave not in agenda:
        print("Contacto no encontrado.")
        return
    agenda[clave]["llamadas"].append((fecha, duracion))  # tupla (fecha, duración)
    print(f"Llamada registrada para {agenda[clave]['nombre']}.")

def mostrar_historial(agenda, clave):
    clave = clave.lower()
    if clave not in agenda:
        print("Contacto no encontrado.")
        return
    datos    = agenda[clave]
    llamadas = datos["llamadas"]
    print(f"Historial de {datos['nombre']}:")
    total = 0
    for fecha, segundos in llamadas:          # desempaquetado de tupla
        minutos = segundos // 60
        segs    = segundos % 60
        print(f"  {fecha}  →  {segundos:4} s  ({minutos} min {segs} s)")
        total += segundos
    mins_total = total // 60
    segs_total = total % 60
    print(f"Total: {total} s ({mins_total} min {segs_total} s)")

registrar_llamada(agenda_llamadas, "ana", "2025-01-25", 90)
mostrar_historial(agenda_llamadas, "ana")
```

**Puntos clave:**
- Las llamadas se almacenan como **tuplas** porque una vez registradas no deben modificarse — son datos históricos.
- `for fecha, segundos in llamadas` usa el **desempaquetado de tupla** directamente en el `for` — elegante y legible.
- La lista `"llamadas"` es mutable (podemos hacer `.append()`), pero cada elemento de la lista es una tupla inmutable.
