## 1. Instalación

## 4. El intérprete de Python

### ¿Cómo funciona Python?

Python es un lenguaje **interpretado**. Esto significa que no necesitas compilar el código (transformarlo en un ejecutable) antes de ejecutarlo. Python lee y ejecuta tu código línea a línea en tiempo real.

```
Tu código (.py o celda)  →  Intérprete de Python  →  Resultado
     print("Hola")       →       lee y ejecuta    →    Hola
```

### Comparación rápida

| | Python (interpretado) | C++ (compilado) |
|--|----------------------|------------------|
| Velocidad de desarrollo | Rápida | Lenta |
| Velocidad de ejecución | Más lenta | Más rápida |
| Facilidad de aprender | Alta | Baja |
| Uso en IA/Data Science | ⭐⭐⭐⭐⭐ | ⭐⭐ |

### Python en la terminal (modo interactivo)

También puedes usar Python directamente en la terminal escribiendo `python` y pulsando Enter. Verás el prompt `>>>` — puedes escribir código línea a línea.

```
$ python
>>> 2 + 2
4
>>> print("Hola")
Hola
>>> exit()
```

## 5. Tu primer programa

### 5.1 `print()` — Mostrar información por pantalla

`print()` es la función más básica de Python. Muestra texto o valores en la salida.

```python
print("Hola, mundo")   # muestra texto
print(42)              # muestra un número
print(3.14)            # muestra un decimal
```

Las palabras entre comillas (`"..."` o `'...'`) se llaman **cadenas de texto** o **strings**.

````python
print("Hola, mundo")
print("Mi nombre es Python")
print("El año es:", 2025)
````

### 5.2 Variables — Guardar información

Una **variable** es como una caja con etiqueta donde guardas un valor para usarlo después.

```
  nombre = "Ana"
  ┌───────┐        ┌─────────┐
  │ nombre│  ────▶ │  "Ana"  │
  └───────┘        └─────────┘
   etiqueta           valor
```

**Reglas para nombrar variables:**
- Solo letras, números y guiones bajos `_`
- No pueden empezar con un número
- Sin espacios (usa `mi_variable` en lugar de `mi variable`)
- Distingue mayúsculas: `nombre` y `Nombre` son variables distintas

**Convención:** usa `snake_case` (palabras separadas por `_`) para los nombres.

````python
# Creamos variables
nombre = "Ana"
edad = 28
altura = 1.65
es_estudiante = True

# Las usamos
print(nombre)
print(edad)
print(altura)
print(es_estudiante)
````

### 5.3 Tipos de datos básicos

Cada valor en Python tiene un **tipo**. Los más fundamentales son:

| Tipo | Nombre en Python | Ejemplo | Para qué se usa |
|------|-----------------|---------|------------------|
| Número entero | `int` | `42`, `-7`, `0` | Contar, indexar |
| Número decimal | `float` | `3.14`, `-0.5` | Medidas, cálculos |
| Texto | `str` | `"hola"`, `'mundo'` | Mensajes, nombres |
| Verdadero/Falso | `bool` | `True`, `False` | Condiciones lógicas |

Puedes saber el tipo de cualquier valor con la función `type()`.

````python
# Comprobamos el tipo de cada variable
print(type(42))          # int
print(type(3.14))        # float
print(type("hola"))      # str
print(type(True))        # bool

# También funciona con variables
edad = 28
print(type(edad))
````

### 5.4 Operadores aritméticos

Python puede hacer cualquier operación matemática:

| Operador | Símbolo | Ejemplo | Resultado |
|----------|---------|---------|----------|
| Suma | `+` | `3 + 2` | `5` |
| Resta | `-` | `10 - 4` | `6` |
| Multiplicación | `*` | `3 * 4` | `12` |
| División | `/` | `10 / 3` | `3.333...` |
| División entera | `//` | `10 // 3` | `3` |
| Módulo (resto) | `%` | `10 % 3` | `1` |
| Potencia | `**` | `2 ** 8` | `256` |

````python
a = 10
b = 3

print("Suma:             ", a + b)
print("Resta:            ", a - b)
print("Multiplicación:   ", a * b)
print("División:         ", a / b)
print("División entera:  ", a // b)
print("Módulo (resto):   ", a % b)
print("Potencia:         ", a ** b)
````

### 5.5 `input()` — Recibir datos del usuario

`input()` detiene el programa y espera a que el usuario escriba algo. Lo que escribe se guarda como un **string**.

```python
nombre = input("¿Cómo te llamas? ")
print("Hola,", nombre)
```

> ⚠️ **Importante:** `input()` siempre devuelve texto (`str`). Si necesitas un número, debes convertirlo con `int()` o `float()`.

```python
edad = int(input("¿Cuántos años tienes? "))
precio = float(input("¿Cuánto cuesta? "))
```

````python
# Programa que saluda al usuario
nombre = input("¿Cómo te llamas? ")
print("¡Hola,", nombre + "! Bienvenido al curso de IA.")
````

````python
# Conversión de tipos
anio_nacimiento = int(input("¿En qué año naciste? "))
edad_aproximada = 2025 - anio_nacimiento
print("Tienes aproximadamente", edad_aproximada, "años.")
````

### 5.6 f-strings — Texto con variables integradas

Una **f-string** es la forma más cómoda de combinar texto y variables. Se escribe poniendo una `f` antes de las comillas, y las variables van entre `{}`.

```python
nombre = "Ana"
edad = 28
print(f"Me llamo {nombre} y tengo {edad} años.")
# → Me llamo Ana y tengo 28 años.
```

También puedes poner expresiones dentro de `{}`:
```python
print(f"El doble de 5 es {5 * 2}")
```

````python
nombre = "Ana"
edad = 28
ciudad = "Madrid"

print(f"Me llamo {nombre}, tengo {edad} años y vivo en {ciudad}.")
print(f"El año que viene tendré {edad + 1} años.")
print(f"El doble de mi edad es {edad * 2}.")
````

### Ejercicio 1 — Calculadora de dos números

**Enunciado:** Escribe un programa que:
1. Pida al usuario dos números.
2. Muestre la suma, la resta, la multiplicación y la división de esos números.

**Ejemplo de salida esperada:**
```
Introduce el primer número: 10
Introduce el segundo número: 4

--- Resultados ---
10.0 + 4.0 = 14.0
10.0 - 4.0 = 6.0
10.0 * 4.0 = 40.0
10.0 / 4.0 = 2.5
```

````python
# ✏️ TU CÓDIGO AQUÍ
````

<details>
<summary>💡 Ver solución — Ejercicio 1</summary>

```python
# Pedimos los dos números al usuario
num1 = float(input("Introduce el primer número: "))
num2 = float(input("Introduce el segundo número: "))

# Calculamos las operaciones
suma        = num1 + num2
resta       = num1 - num2
producto    = num1 * num2
division    = num1 / num2

# Mostramos los resultados
print("\n--- Resultados ---")
print(f"{num1} + {num2} = {suma}")
print(f"{num1} - {num2} = {resta}")
print(f"{num1} * {num2} = {producto}")
print(f"{num1} / {num2} = {division}")
```

**Puntos clave:**
- Usamos `float()` para que el usuario pueda introducir decimales.
- `\n` dentro de un string imprime una línea en blanco.
- Las f-strings nos permiten mostrar la operación y el resultado en la misma línea de forma legible.
</details>

---
### Ejercicio 2 — División con resto

**Enunciado:** Modifica la calculadora para que también muestre:
- La **división entera** (cuántas veces cabe el segundo número en el primero).
- El **resto** de esa división.

**Ejemplo de salida esperada:**
```
Introduce el primer número: 17
Introduce el segundo número: 5

17 entre 5 cabe 3 veces y sobran 2
```

````python
# ✏️ TU CÓDIGO AQUÍ
````

<details>
<summary>💡 Ver solución — Ejercicio 2</summary>

```python
num1 = int(input("Introduce el primer número: "))
num2 = int(input("Introduce el segundo número: "))

division_entera = num1 // num2
resto           = num1 % num2

print(f"\n{num1} entre {num2} cabe {division_entera} veces y sobran {resto}")
```

**Puntos clave:**
- Usamos `int()` en lugar de `float()` porque la división con resto tiene más sentido con enteros.
- `//` es la división entera (descarta los decimales).
- `%` devuelve el resto de la división.
</details>

### Ejercicio 3 — Calculadora de área y perímetro

**Enunciado:** Escribe un programa que pida el **radio de un círculo** y calcule:
- Su **área** (fórmula: π × r²)
- Su **perímetro** (fórmula: 2 × π × r)

Usa `3.14159` como valor de π (o importa `math.pi` si quieres ser más preciso).

**Ejemplo de salida esperada:**
```
Introduce el radio del círculo: 5

Círculo con radio 5.0:
  Área:      78.53975
  Perímetro: 31.4159
```

````python
# ✏️ TU CÓDIGO AQUÍ
# Pista: para usar pi de la librería math, escribe al inicio:
# import math
# y luego usa math.pi en lugar de 3.14159
````

<details>
<summary>💡 Ver solución — Ejercicio 3</summary>

```python
import math

radio = float(input("Introduce el radio del círculo: "))

area      = math.pi * radio ** 2
perimetro = 2 * math.pi * radio

print(f"\nCírculo con radio {radio}:")
print(f"  Área:      {area:.5f}")
print(f"  Perímetro: {perimetro:.5f}")
```

**Puntos clave:**
- `import math` carga la librería matemática de Python. Las librerías son colecciones de funciones ya escritas que puedes reutilizar.
- `math.pi` es el valor de π con máxima precisión.
- `{area:.5f}` dentro de una f-string formatea el número con 5 decimales. El formato es `{valor:.Nf}` donde N es el número de decimales.
</details>

### Ejercicio 4 (reto) — Calculadora de propinas

**Enunciado:** En un restaurante, escribe un programa que:
1. Pida el **total de la cuenta** en euros.
2. Pida el **porcentaje de propina** que quiere dejar el usuario (ej: 10 para 10%).
3. Muestre:
   - El importe de la propina.
   - El total a pagar (cuenta + propina).
   - Si hay 4 personas, **¿cuánto paga cada una?**

**Ejemplo de salida esperada:**
```
Total de la cuenta (€): 85.50
Porcentaje de propina (%): 15
Número de personas: 4

--- Resumen ---
Cuenta:          85.50 €
Propina (15%):   12.82 €
Total:           98.32 €
Por persona:     24.58 €
```

````python
# ✏️ TU CÓDIGO AQUÍ
````

<details>
<summary>💡 Ver solución — Ejercicio 4</summary>

```python
cuenta    = float(input("Total de la cuenta (€): "))
porcentaje = float(input("Porcentaje de propina (%): "))
personas  = int(input("Número de personas: "))

propina    = cuenta * (porcentaje / 100)
total      = cuenta + propina
por_persona = total / personas

print("\n--- Resumen ---")
print(f"Cuenta:          {cuenta:.2f} €")
print(f"Propina ({porcentaje:.0f}%):   {propina:.2f} €")
print(f"Total:           {total:.2f} €")
print(f"Por persona:     {por_persona:.2f} €")
```

**Puntos clave:**
- `porcentaje / 100` convierte el porcentaje en un factor multiplicador (15% → 0.15).
- `{valor:.2f}` formatea con exactamente 2 decimales, ideal para monedas.
- `{porcentaje:.0f}` muestra el porcentaje sin decimales aunque sea `float`.
</details>
