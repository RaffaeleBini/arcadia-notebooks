## 1. Operadores aritméticos

Son los operadores matemáticos clásicos. Ya los vimos en la Clase 1, aquí los profundizamos.

| Operador | Nombre | Ejemplo | Resultado |
|----------|--------|---------|----------|
| `+` | Suma | `7 + 3` | `10` |
| `-` | Resta | `7 - 3` | `4` |
| `*` | Multiplicación | `7 * 3` | `21` |
| `/` | División real | `7 / 2` | `3.5` |
| `//` | División entera | `7 // 2` | `3` |
| `%` | Módulo (resto) | `7 % 2` | `1` |
| `**` | Potencia | `2 ** 8` | `256` |
| `-x` | Negación unaria | `-7` | `-7` |

### Relación entre `/`, `//` y `%`

Están relacionados por esta fórmula:
```
a = (a // b) * b + (a % b)
```
Ejemplo: `17 = (17 // 5) * 5 + (17 % 5)` → `17 = 3 * 5 + 2` → `17 = 17` ✓

````python
a, b = 17, 5   # asignación múltiple: a = 17, b = 5 en una sola línea

print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} / {b}  = {a / b}")    # siempre float
print(f"{a} // {b} = {a // b}")   # cociente entero
print(f"{a} % {b}  = {a % b}")    # resto
print(f"{a} ** {b} = {a ** b}")   # potencia

# Verificamos la relación entre //, % y el original
verificacion = (a // b) * b + (a % b)
print(f"\nVerificación: ({a}//{b})*{b} + ({a}%{b}) = {verificacion} == {a}")
````

### 1.1 Precedencia de operadores (orden de evaluación)

Cuando una expresión tiene varios operadores, Python los evalúa en un orden concreto — igual que en matemáticas:

| Prioridad | Operador(es) | Nombre |
|-----------|-------------|--------|
| 1 (mayor) | `()` | Paréntesis |
| 2 | `**` | Potencia |
| 3 | `-x` | Negación unaria |
| 4 | `*`, `/`, `//`, `%` | Multiplicación y división |
| 5 (menor) | `+`, `-` | Suma y resta |

> **Regla de oro:** cuando tengas dudas, usa **paréntesis**. Hacen el código más legible y evitan errores.

````python
# Sin paréntesis: Python aplica las reglas de precedencia
print(2 + 3 * 4)       # 14, no 20 (primero * luego +)
print(10 - 2 ** 3)     # 2,  no 512 (primero ** luego -)
print(10 / 2 + 3)      # 8.0 (primero / luego +)

print("---")

# Con paréntesis: controlamos el orden
print((2 + 3) * 4)     # 20
print((10 - 2) ** 3)   # 512
print(10 / (2 + 3))    # 2.0

print("---")

# Ejemplo real: fórmula del interés compuesto   C * (1 + r)^t
capital = 1000
tasa    = 0.05    # 5% anual
años    = 10

# MAL (sin paréntesis, la precedencia cambia el resultado)
mal  = capital * 1 + tasa ** años
# BIEN
bien = capital * (1 + tasa) ** años

print(f"Sin paréntesis (MAL): {mal:.2f} €")
print(f"Con paréntesis (BIEN): {bien:.2f} €")
````

### 1.2 Operadores de asignación compuesta

Son atajos para modificar una variable aplicándole una operación:

| Operador | Equivale a | Ejemplo |
|----------|-----------|--------|
| `x += 5` | `x = x + 5` | Suma 5 a x |
| `x -= 3` | `x = x - 3` | Resta 3 a x |
| `x *= 2` | `x = x * 2` | Multiplica x por 2 |
| `x /= 4` | `x = x / 4` | Divide x entre 4 |
| `x //= 3` | `x = x // 3` | División entera de x entre 3 |
| `x %= 7` | `x = x % 7` | Resto de x entre 7 |
| `x **= 2` | `x = x ** 2` | x elevado al cuadrado |

````python
puntos = 100
print(f"Inicio:            {puntos}")

puntos += 50      # ganas 50 puntos
print(f"Tras ganar 50:     {puntos}")

puntos -= 30      # pierdes 30 puntos
print(f"Tras perder 30:    {puntos}")

puntos *= 2       # puntos dobles
print(f"Puntos dobles:     {puntos}")

puntos //= 3      # divides entre 3 equipos
print(f"Divididos en 3:    {puntos}")
````

## 2. Operadores de comparación

Comparan dos valores y **siempre devuelven un booleano** (`True` o `False`).

| Operador | Significado | Ejemplo | Resultado |
|----------|------------|---------|----------|
| `==` | Igual a | `5 == 5` | `True` |
| `!=` | Distinto de | `5 != 3` | `True` |
| `>` | Mayor que | `5 > 3` | `True` |
| `<` | Menor que | `5 < 3` | `False` |
| `>=` | Mayor o igual | `5 >= 5` | `True` |
| `<=` | Menor o igual | `5 <= 4` | `False` |

> ⚠️ **Error clásico:** confundir `=` (asignación) con `==` (comparación).
> ```python
> x = 5      # asigna el valor 5 a x
> x == 5     # pregunta si x vale 5 → True
> ```

````python
a = 10
b = 7

print(f"{a} == {b}  →  {a == b}")
print(f"{a} != {b}  →  {a != b}")
print(f"{a} >  {b}  →  {a > b}")
print(f"{a} <  {b}  →  {a < b}")
print(f"{a} >= {b}  →  {a >= b}")
print(f"{a} <= {b}  →  {a <= b}")
````

````python
# Comparar strings: Python compara letra a letra en orden alfabético (Unicode)
print("apple" == "apple")   # True
print("apple" != "banana")  # True
print("apple" < "banana")   # True  (a < b en el alfabeto)
print("Zebra" < "apple")    # True  (mayúsculas tienen código menor que minúsculas)

print("---")

# Comparaciones encadenadas — Python las soporta de forma elegante
x = 5
print(1 < x < 10)     # True  (equivale a 1 < x AND x < 10)
print(0 < x <= 5)     # True
print(1 < x < 4)      # False
````

## 3. Operadores lógicos

Combinan expresiones booleanas. Son tres: `and`, `or`, `not`.

### `and` — Las dos condiciones deben ser verdaderas

```
True  and True   → True
True  and False  → False
False and True   → False
False and False  → False
```

### `or` — Al menos una condición debe ser verdadera

```
True  or True   → True
True  or False  → True
False or True   → True
False or False  → False
```

### `not` — Invierte el valor

```
not True   → False
not False  → True
```

````python
# Tablas de verdad completas
print("=== AND ===")
for a in [True, False]:
    for b in [True, False]:
        print(f"  {str(a):<6} and {str(b):<6} → {a and b}")

print("\n=== OR ===")
for a in [True, False]:
    for b in [True, False]:
        print(f"  {str(a):<6} or  {str(b):<6} → {a or b}")

print("\n=== NOT ===")
for a in [True, False]:
    print(f"  not {str(a):<6} → {not a}")
````

````python
# Casos prácticos
edad          = 20
tiene_carnet  = True
llueve        = False
tiene_paraguas = True

# AND: las dos condiciones deben cumplirse
puede_conducir = edad >= 18 and tiene_carnet
print(f"¿Puede conducir?      {puede_conducir}")

# OR: con que se cumpla una es suficiente
sale_sin_mojarse = not llueve or tiene_paraguas
print(f"¿Sale sin mojarse?    {sale_sin_mojarse}")

# NOT: invertir una condición
no_llueve = not llueve
print(f"¿No llueve?           {no_llueve}")

# Combinación compleja
puede_ir_al_parque = edad >= 18 and (not llueve or tiene_paraguas)
print(f"¿Puede ir al parque?  {puede_ir_al_parque}")
````

### 3.1 Evaluación en cortocircuito

Python es inteligente: en una expresión `and` u `or`, deja de evaluar en cuanto puede determinar el resultado.

- `False and (...)` → el resultado ya es `False`, el segundo operando **no se evalúa**.
- `True or (...)` → el resultado ya es `True`, el segundo operando **no se evalúa**.

Esto es importante para evitar errores — por ejemplo, comprobar que un número no es cero **antes** de dividir:

````python
# Sin cortocircuito, la división por cero daría un error
# Con and, si b == 0 la segunda parte no se evalúa
a = 10
b = 0

# Forma segura gracias al cortocircuito:
resultado = b != 0 and a / b > 2
print(f"b != 0 and a/b > 2  →  {resultado}")  # False (sin error)

# También funciona con or:
nombre = ""
nombre_mostrar = nombre or "Usuario anónimo"   # si nombre es vacío (falsy), usa el default
print(f"Nombre a mostrar: {nombre_mostrar}")
````

## 4. Operadores especiales

### 4.1 Operador de identidad: `is` y `is not`

`is` comprueba si dos variables apuntan al **mismo objeto en memoria** (no si tienen el mismo valor).

```python
a == b    # ¿tienen el mismo valor?
a is b    # ¿son el mismo objeto en memoria?
```

El uso más común es comprobar si algo es `None`:
```python
resultado is None       # correcto
resultado == None       # funciona pero no es idiomático
```

### 4.2 Operador de pertenencia: `in` y `not in`

`in` comprueba si un elemento está dentro de una secuencia (string, lista, etc.).

````python
# Operador 'in' con strings
frase = "Python es genial"
print("Python" in frase)       # True
print("Java" in frase)         # False
print("Java" not in frase)     # True

print("---")

# is None — forma correcta de comprobar ausencia de valor
valor = None
print(valor is None)           # True
print(valor is not None)       # False

print("---")

# Diferencia entre == e is
a = [1, 2, 3]
b = [1, 2, 3]   # mismos valores, objeto diferente
c = a           # misma referencia

print(f"a == b  →  {a == b}")   # True  (mismo contenido)
print(f"a is b  →  {a is b}")   # False (objetos distintos en memoria)
print(f"a is c  →  {a is c}")   # True  (c apunta al mismo objeto que a)
````

## 5. Expresiones

Una **expresión** es cualquier combinación de valores, variables y operadores que Python puede **evaluar** para producir un resultado.

```python
2 + 3              # expresión aritmética → 5
x > 0              # expresión de comparación → True o False
a and b or not c   # expresión lógica → True o False
"hola" + nombre    # expresión con strings → un string nuevo
```

Una **sentencia** es una instrucción completa que Python ejecuta pero que no necesariamente produce un valor:
```python
x = 5              # sentencia de asignación
print(x)           # sentencia de llamada a función
```

### Precedencia completa de operadores

Cuando mezclas operadores aritméticos, de comparación y lógicos:

| Prioridad | Operadores |
|-----------|------------|
| 1 (mayor) | `()` |
| 2 | `**` |
| 3 | `-x` (negación) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `is`, `in` |
| 7 | `not` |
| 8 | `and` |
| 9 (menor) | `or` |

````python
# Python evalúa: aritmética → comparación → lógica
x = 8

# Paso a paso: primero 3*2=6, luego 8>6 → True, luego True and True → True
resultado = x > 3 * 2 and x < 100
print(f"x > 3*2 and x < 100  →  {resultado}")

# not tiene menor precedencia que las comparaciones pero mayor que and
print(not x == 8)       # False  (primero x==8 → True, luego not True → False)
print(not x == 8 or True)  # True

print("---")

# Expresión condicional (ternario) — forma compacta de if/else
# valor_si_verdad  if  condición  else  valor_si_falso
edad = 20
estado = "mayor de edad" if edad >= 18 else "menor de edad"
print(f"Con {edad} años eres {estado}")

# Otro ejemplo
numero = -5
signo = "positivo" if numero >= 0 else "negativo"
print(f"{numero} es {signo}")
````

### 5.1 El operador morsa `:=` (Python 3.8+)

El operador `:=` (llamado *walrus* por su parecido con una morsa 🦭) permite **asignar y usar un valor en la misma expresión**. Es útil para evitar calcular algo dos veces.

````python
import math

# Sin walrus: calculamos y luego comprobamos
radio = 7
area = math.pi * radio ** 2
if area > 100:
    print(f"Área grande: {area:.2f}")

# Con walrus: calculamos y comprobamos en un paso
if (area := math.pi * radio ** 2) > 100:
    print(f"Área grande (walrus): {area:.2f}")
````

### Ejercicio 5 (reto) — Calculadora científica

**Enunciado:** Escribe una calculadora que resuelva la **ecuación de segundo grado**:

```
ax² + bx + c = 0
```

El programa debe:
1. Pedir los coeficientes `a`, `b`, `c`.
2. Calcular el **discriminante**: `D = b² - 4ac`
3. Usando operadores de comparación y lógicos, determinar:
   - Si `D > 0`: dos soluciones reales
   - Si `D == 0`: una solución (raíz doble)
   - Si `D < 0`: sin soluciones reales
4. Calcular y mostrar las soluciones cuando existan.

Fórmula: `x = (-b ± √D) / (2a)`

**Ejemplo de salida esperada:**
```
Coeficiente a: 1
Coeficiente b: -5
Coeficiente c: 6

Discriminante D = 1
Dos soluciones reales:
  x1 = 3.0
  x2 = 2.0
```

```python
import math

a = float(input("Coeficiente a: "))
b = float(input("Coeficiente b: "))
c = float(input("Coeficiente c: "))

D = b**2 - 4*a*c

print(f"\nDiscriminante D = {D}")

tiene_dos_soluciones = D > 0
tiene_una_solucion   = D == 0
sin_solucion_real    = D < 0

if tiene_dos_soluciones:
    x1 = (-b + math.sqrt(D)) / (2*a)
    x2 = (-b - math.sqrt(D)) / (2*a)
    print("Dos soluciones reales:")
    print(f"  x1 = {x1}")
    print(f"  x2 = {x2}")
elif tiene_una_solucion:
    x = -b / (2*a)
    print(f"Una solución (raíz doble): x = {x}")
else:
    print("Sin soluciones reales (D < 0)")
```

**Puntos clave:**
- El discriminante `D = b**2 - 4*a*c` es el punto clave: su signo determina el tipo de solución.
- Guardamos los booleanos en variables descriptivas antes del `if` — facilita la lectura.
- Este ejercicio anticipa los condicionales `if/elif/else` de la próxima clase — ya tienes una primera visión de cómo funcionan.
- `math.sqrt()` solo acepta números no negativos; por eso comprobamos que `D >= 0` antes de usarla.
