## 1. Perché la OOP?

Immagina di dover gestire 100 dipendenti. Senza OOP:

```python
# Senza OOP — dati sparsi, difficile da mantenere
nome_1     = "Ana"
stipendio_1 = 2500
nome_2     = "Luis"
stipendio_2 = 3000

def calcola_bonus(stipendio):

```

Con OOP:

```python
# Con OOP — dati e comportamento insieme, facile da scalare

    def calcular_bonus(self):
        return self.salario * 0.10

ana  = Empleado("Ana",  2500)
luis = Empleado("Luis", 3000)
ana.calcular_bonus()    # 250
```

## 2. Classi e oggetti

Una **classe** è il modello — definisce la struttura.
Un **oggetto** (o istanza) è ogni copia concreta creata a partire da quel modello.

```
Classe Empleado
  ├── attributi: nome, stipendio, dipartimento
  └── metodi:  saluta(), calcola_bonus()
        │
        ├── Oggetto: ana  (nome="Ana",  stipendio=2500)
        ├── Oggetto: luis (nome="Luis", stipendio=3000)
        └── Oggetto: marta(nome="Marta",stipendio=2800)
```

Analogia: la classe è lo **stampo** per fare biscotti; ogni biscotto è un oggetto distinto ma tutti hanno la stessa forma.

````python
# La classe più semplice possibile
class Perro:
    pass   # 'pass' perché non abbiamo ancora aggiunto nulla

# Creare oggetti (istanziare la classe)
rex   = Perro()
laika = Perro()

print(type(rex))          # <class '__main__.Perro'>
print(isinstance(rex, Perro))  # True
print(rex is laika)       # False — sono oggetti distinti

# Possiamo aggiungere attributi «al volo» (non è il modo corretto, ma mostra il concetto)
rex.nombre = "Rex"
rex.edad   = 3
print(f"{rex.nombre} ha {rex.edad} anni")
````

## 3. Il metodo `__init__` — Costruttore

`__init__` è il metodo che Python chiama automaticamente quando crei un oggetto con `Classe()`. La sua funzione è inizializzare gli attributi dell'oggetto.

```python
class MiaClasse:
    def __init__(self, param1, param2):
        self.attributo1 = param1
        self.attributo2 = param2
```

**Cos'è `self`?**

`self` è un riferimento all'**oggetto stesso** che si sta creando. Python lo passa automaticamente — tu non lo passi mai quando chiami il metodo. La convenzione di chiamarlo `self` non è obbligatoria ma è universale.

```
Empleado("Ana", 2500)
      │
      └──► Python chiama __init__(self=<nuovo oggetto>, nombre="Ana", salario=2500)
```

````python
class Perro:
    def __init__(self, nombre, raza, edad):
        self.nombre = nombre
        self.raza   = raza
        self.edad   = edad

# Quando istanzi, Python chiama __init__ automaticamente
rex   = Perro("Rex",   "Pastore Tedesco", 3)
laika = Perro("Laika", "Husky",           5)

print(f"{rex.nombre}   — {rex.raza},   {rex.edad} anni")
print(f"{laika.nombre} — {laika.raza}, {laika.edad} anni")

# Ogni oggetto ha i propri dati
rex.edad = 4   # modificare un attributo
print(f"\nEtà di Rex aggiornata:   {rex.edad}")
print(f"Età di Laika invariata:   {laika.edad}")
````

````python
# Parametri con valori di default in __init__
class Perro:
    def __init__(self, nombre, raza="Sconosciuta", edad=0):
        self.nombre = nombre
        self.raza   = raza
        self.edad   = edad

    def descripcion(self):
        return f"{self.nombre} ({self.raza}, {self.edad} anni)"

perro1 = Perro("Max")
perro2 = Perro("Nala", "Labrador")
perro3 = Perro("Thor", "Rottweiler", 2)

for p in [perro1, perro2, perro3]:
    print(p.descripcion())
````

## 4. Attributi

Python ha due tipi principali di attributi:

| Tipo | Dove si definisce | A chi appartiene | Accesso |
|------|-----------------|-------------------|--------|
| **Di istanza** | Dentro `__init__` con `self.x` | A ogni oggetto | `oggetto.x` |
| **Di classe** | Dentro la classe, fuori dai metodi | Alla classe (condiviso) | `Classe.x` o `oggetto.x` |

````python
class Contador:
    total_instancias = 0   # ← attributo di CLASSE (condiviso)

    def __init__(self, nombre):
        self.nombre = nombre               # ← attributo di ISTANZA
        Contador.total_instancias += 1     # aggiorna il contatore condiviso

c1 = Contador("primo")
c2 = Contador("secondo")
c3 = Contador("terzo")

print(f"Istanze create: {Contador.total_instancias}")     # 3 (attributo di classe)
print(f"Nome di c1: {c1.nombre}")                          # primo (attributo di istanza)
print(f"Nome di c2: {c2.nombre}")                          # secondo

# L'attributo di classe è condiviso — se cambia, cambia per tutti
print(f"\nDall'istanza c1: {c1.total_instancias}")   # 3
print(f"Dall'istanza c2: {c2.total_instancias}")     # 3
````

````python
# ⚠️ Attenzione: se assegni sull'istanza, crei un nuovo attributo su quell'istanza
# e quello di classe resta intatto

class Config:
    version = "1.0"   # di classe

cfg1 = Config()
cfg2 = Config()

cfg1.version = "2.0"   # crea un attributo di ISTANZA in cfg1

print(f"cfg1.version:     {cfg1.version}")    # 2.0 (istanza)
print(f"cfg2.version:     {cfg2.version}")    # 1.0 (classe)
print(f"Config.version:   {Config.version}")  # 1.0 (classe, intatta)

# Verificare se l'attributo è di istanza o di classe
print(f"\n'version' in cfg1.__dict__: {'version' in cfg1.__dict__}")  # True
print(f"'version' in cfg2.__dict__: {'version' in cfg2.__dict__}")  # False
````

## 5. Metodi

Un **metodo** è una funzione definita dentro una classe. Ce ne sono tre tipi:

| Tipo | Decoratore | Primo parametro | Quando usarlo |
|------|-----------|-----------------|---------------|
| Di istanza | (nessuno) | `self` | Accede o modifica i dati dell'oggetto |
| Di classe | `@classmethod` | `cls` | Lavora con attributi di classe; costruttori alternativi |
| Statico | `@staticmethod` | (nessuno) | Utilità collegata alla classe ma senza bisogno di `self` né `cls` |

````python
class Rectangulo:
    # ─── Attributo di classe ───────────────────────────────
    unidad = "cm"

    # ─── Costruttore ─────────────────────────────────────
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto  = alto

    # ─── Metodi di istanza ────────────────────────────
    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)

    def es_cuadrado(self):
        return self.ancho == self.alto

    def escalar(self, factor):
        """Modifica l'oggetto sul posto."""
        self.ancho *= factor
        self.alto  *= factor

    # ─── Metodo di classe (costruttore alternativo) ───────
    @classmethod
    def cuadrado(cls, lado):
        """Crea un rettangolo quadrato senza ripetere la logica."""
        return cls(lado, lado)

    @classmethod
    def cambiar_unidad(cls, nueva_unidad):
        cls.unidad = nueva_unidad

    # ─── Metodo statico ─────────────────────────────────
    @staticmethod
    def es_valido(ancho, alto):
        """Verifica che le misure siano positive — non serve self né cls."""
        return ancho > 0 and alto > 0


# Utilizzo
r = Rectangulo(6, 4)
print(f"Area:        {r.area()} {Rectangulo.unidad}²")
print(f"Perimetro:   {r.perimetro()} {Rectangulo.unidad}")
print(f"È un quadrato?   {r.es_cuadrado()}")

q = Rectangulo.cuadrado(5)          # costruttore alternativo
print(f"\nQuadrato:    {q.ancho}x{q.alto}, area={q.area()}")
print(f"È un quadrato?   {q.es_cuadrado()}")

print(f"\n0x4 è valido? {Rectangulo.es_valido(0, 4)}")   # metodo statico
print(f"5x3 è valido? {Rectangulo.es_valido(5, 3)}")

r.escalar(2)
print(f"\nDopo aver scalato x2: {r.ancho}x{r.alto}, area={r.area()}")
````

## 6. Metodi speciali (`dunder`)

I metodi con doppio underscore all'inizio e alla fine (`__metodo__`) si chiamano **dunder** (double underscore). Python li chiama automaticamente in risposta a operatori o funzioni built-in.

| Metodo | Viene chiamato quando… | Esempio |
|--------|------------------|---------|
| `__init__` | Crei un oggetto | `Classe()` |
| `__str__` | Converti in string leggibile | `str(obj)`, `print(obj)` |
| `__repr__` | Rappresentazione per il debugging | `repr(obj)`, nel REPL |
| `__eq__` | Confronti con `==` | `obj1 == obj2` |
| `__lt__` | Confronti con `<` | `obj1 < obj2` |
| `__len__` | Chiami `len()` | `len(obj)` |
| `__add__` | Usi `+` | `obj1 + obj2` |
| `__contains__` | Usi `in` | `x in obj` |

````python
class Libro:
    def __init__(self, titulo, autor, paginas):
        self.titulo  = titulo
        self.autor   = autor
        self.paginas = paginas

    def __str__(self):
        """Rappresentazione leggibile per l'utente."""
        return f"«{self.titulo}» di {self.autor} ({self.paginas} pagg.)"

    def __repr__(self):
        """Rappresentazione tecnica per il debugging."""
        return f"Libro(titulo={self.titulo!r}, autor={self.autor!r}, paginas={self.paginas})"

    def __eq__(self, otro):
        """Due libri sono uguali se hanno lo stesso titolo e lo stesso autore."""
        if not isinstance(otro, Libro):
            return NotImplemented
        return self.titulo == otro.titulo and self.autor == otro.autor

    def __lt__(self, otro):
        """Ordinare per numero di pagine."""
        return self.paginas < otro.paginas

    def __len__(self):
        return self.paginas


b1 = Libro("1984",          "George Orwell",  328)
b2 = Libro("Dune",          "Frank Herbert",  604)
b3 = Libro("1984",          "George Orwell",  328)  # uguale a b1

print(str(b1))          # __str__
print(repr(b2))         # __repr__
print(f"b1 == b3: {b1 == b3}")  # __eq__   → True
print(f"b1 == b2: {b1 == b2}")  # __eq__   → False
print(f"b1 < b2:  {b1 < b2}")   # __lt__   → True (328 < 604)
print(f"len(b2):  {len(b2)}")   # __len__

# sorted() usa automaticamente __lt__
biblioteca = [b2, b1, Libro("Il Piccolo Principe", "Saint-Exupéry", 96)]
for libro in sorted(biblioteca):
    print(f"  {len(libro):>4} pagg — {libro}")
````

## 7. Incapsulamento — Attributi privati e `@property`

L'incapsulamento controlla quali parti dell'oggetto sono accessibili dall'esterno. Python usa convenzioni di nomenclatura:

| Convenzione | Esempio | Significato |
|------------|---------|-------------|
| Senza prefisso | `self.nombre` | Pubblico — accessibile da qualsiasi punto |
| Un underscore | `self._nombre` | Protetto — convenzione «non toccare dall'esterno» |
| Due underscore | `self.__nombre` | Privato — Python applica il *name mangling* |

> Python non impedisce l'accesso tecnicamente, ma i prefissi sono segnali chiari per gli altri programmatori (e per te stesso in futuro).

````python
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular  = titular           # pubblico
        self._saldo   = saldo_inicial     # protetto (convenzione)
        self.__pin    = "0000"            # privato (name mangling)

    # @property trasforma un metodo in un attributo di sola lettura
    @property
    def saldo(self):
        return self._saldo

    # @saldo.setter permette di assegnare con validazione
    @saldo.setter
    def saldo(self, valor):
        if valor < 0:
            raise ValueError("Il saldo non può essere negativo")
        self._saldo = valor

    def depositar(self, cantidad):
        if cantidad <= 0:
            raise ValueError("La quantità deve essere positiva")
        self._saldo += cantidad
        return self._saldo

    def retirar(self, cantidad):
        if cantidad > self._saldo:
            raise ValueError("Fondi insufficienti")
        self._saldo -= cantidad
        return self._saldo

    def __str__(self):
        return f"Conto di {self.titular}: {self._saldo:.2f} €"


cuenta = CuentaBancaria("Ana", 1000)
print(cuenta)

cuenta.depositar(500)
print(f"Dopo il deposito:  {cuenta.saldo:.2f} €")   # via @property

cuenta.retirar(200)
print(f"Dopo il prelievo:  {cuenta.saldo:.2f} €")

# Il setter valida
try:
    cuenta.saldo = -100
except ValueError as e:
    print(f"Errore: {e}")

# L'attributo privato __pin applica il name mangling
try:
    print(cuenta.__pin)    # AttributeError
except AttributeError:
    print("Non si può accedere a __pin direttamente")
    print(f"(esiste come: {cuenta._CuentaBancaria__pin})")  # name mangling
````

````python
# @property per attributi calcolati (senza setter — di sola lettura)
import math

class Circulo:
    def __init__(self, radio):
        self.radio = radio

    @property
    def area(self):
        return math.pi * self.radio ** 2

    @property
    def perimetro(self):
        return 2 * math.pi * self.radio

    @radio.setter
    def radio(self, valor):
        if valor <= 0:
            raise ValueError("Il raggio deve essere positivo")
        self._radio = valor

    @property
    def radio(self):
        return self._radio

    def __str__(self):
        return f"Cerchio(r={self.radio}, area={self.area:.2f}, per={self.perimetro:.2f})"

c = Circulo(5)
print(c)
c.radio = 10
print(c)

# area e perimetro vengono ricalcolati automaticamente al cambiare del raggio
try:
    c.area = 100   # TypeError — non ha setter
except AttributeError as e:
    print(f"Non si può assegnare direttamente: {e}")
````

### Esercizio 1 — La classe base `Empleado`

**Consegna:** Crea la classe `Empleado` con:

- **Attributi di classe:** `empresa = "MiEmpresa S.L."` e `_contador_ids = 0` (si incrementa a ogni istanza).
- **`__init__`** che riceve `nombre`, `apellido`, `departamento` e `salario`. Assegna un `id` unico autoincrementale. Salva `salario` in `_salario` (privato).

- **`__repr__`** tecnico: `"Empleado(id='E001', nombre='Ana', apellido='García', ...)"`.

L'`id` viene formattato come `"E" + numero con zeri a sinistra`, ad esempio `E001`, `E002`.

```python
class Empleado:
    # ─── Attributi di classe ───────────────────────────────
    empresa       = "MiEmpresa S.L."
    _contador_ids = 0

    # ─── Costruttore ─────────────────────────────────────
    def __init__(self, nombre, apellido, departamento, salario):
        Empleado._contador_ids += 1
        self.id           = f"E{Empleado._contador_ids:03d}"
        self.nombre       = nombre.strip().title()
        self.apellido     = apellido.strip().title()
        self.departamento = departamento
        self.salario      = salario          # usa il setter con validazione

    # ─── Proprietà salario ───────────────────────────────
    @property
    def salario(self):
        return self._salario

    @salario.setter
    def salario(self, valor):
        if not isinstance(valor, (int, float)) or valor < 0:
            raise ValueError(f"Stipendio non valido: {valor}. Deve essere un numero ≥ 0")
        self._salario = float(valor)

    # ─── Metodi dunder ──────────────────────────────────
    def __str__(self):
        return (
            f"[{self.id}] {self.nombre} {self.apellido} "
            f"— {self.departamento} ({self._salario:,.2f} €)"
        )

    def __repr__(self):
        return (
            f"Empleado(id={self.id!r}, nombre={self.nombre!r}, "
            f"apellido={self.apellido!r}, departamento={self.departamento!r}, "
            f"salario={self._salario})"
        )

    def __eq__(self, otro):
        if not isinstance(otro, Empleado):
            return NotImplemented
        return self.id == otro.id


# Prova
e1 = Empleado("ana", "garcía",  "Desarrollo", 2500)
e2 = Empleado("Luis", "Pérez",  "Marketing",  3100)
e3 = Empleado("Marta", "López", "Desarrollo", 2800)

print(e1)
print(e2)
print(repr(e3))
print(f"\ne1 == e2: {e1 == e2}")   # False
print(f"e1 == e1: {e1 == e1}")   # True

try:
    e1.salario = -500
except ValueError as e:
    print(f"\nErrore stipendio: {e}")
```

**Punti chiave:**
- `Empleado._contador_ids += 1` dentro `__init__` garantisce che l'id venga assegnato prima di qualsiasi altra operazione.
- `self.salario = salario` (senza underscore) nell'`__init__` passa attraverso il setter — la validazione viene applicata già in fase di creazione.
- `.title()` normalizza il nome indipendentemente da come lo scrive l'utente.
- `__eq__` restituisce `NotImplemented` (non `False`) quando l'altro oggetto non è un `Empleado` — così Python può tentare il confronto inverso.

    def __repr__(self):
        return f"Empleado(id={self.id!r}, nombre={self.nombre!r}, apellido={self.apellido!r}, salario={self._salario})"

    def __eq__(self, otro):
        if not isinstance(otro, Empleado):
            return NotImplemented
        return self.id == otro.id

    def __lt__(self, otro):
        return self._salario < otro._salario


# Reimposta il contatore
Empleado._contador_ids = 0

e1 = Empleado("ana",   "garcía",  "Desarrollo", 2500)

print("\n--- Ordinati per stipendio ---")
for emp in sorted([e2, e1, e3]):
    print(f"  {emp}")

**Punti chiave:**

    @classmethod
    def total_empleados(cls):
        return cls._contador_ids

    @classmethod
    def crear_desde_dict(cls, datos):
        """Costruttore alternativo da un dizionario."""
        return cls(
            datos["nombre"],
            datos["apellido"],
            datos["departamento"],
            datos["salario"]
        )

    @classmethod
    def crear_desde_csv_linea(cls, linea):
        """Costruttore alternativo da una riga CSV."""
        partes = [p.strip() for p in linea.split(",")]
        if len(partes) != 4:
            raise ValueError(f"Riga CSV non valida: attesi 4 campi, arrivati {len(partes)}")
        nombre, apellido, departamento, salario_str = partes
        try:
            salario = float(salario_str)
        except ValueError:
            raise ValueError(f"Lo stipendio '{salario_str}' non è un numero valido")
        return cls(nombre, apellido, departamento, salario)

    @staticmethod
    def es_salario_valido(salario):
        """Restituisce True se lo stipendio è un numero >= 0."""
        return isinstance(salario, (int, float)) and salario >= 0

    @staticmethod
    def formatear_nombre(nombre):
        """Capitalizza correttamente nomi composti."""
        return " ".join(parte.capitalize() for parte in nombre.split())

- `crear_desde_csv_linea` converte lo stipendio con `float()` dentro un `try/except` — non fidarti mai di dati esterni.
- `@staticmethod` non riceve né `self` né `cls` — è una funzione normale raggruppata dentro la classe per motivi organizzativi.
- `" ".join(parte.capitalize() for parte in nombre.split())` gestisce nomi composti come `"ANA MARÍA"` meglio di `.title()` (`.title()` gestirebbe bene `"Ana María"`, ma fallirebbe con `"O'Brien"` rispetto al corretto `"O'Brien"`).

### Esercizio 4 — La classe `Departamento`

**Consegna:** Crea una classe `Departamento` che gestisca una raccolta di dipendenti:

- **`__init__(nombre)`** — inizializza con nome e lista vuota di dipendenti.
- **`añadir(empleado)`** — aggiunge un `Empleado`. Lancia `TypeError` se non è istanza di `Empleado`.
- **`eliminar(id_empleado)`** — elimina per id. Lancia `ValueError` se non esiste.
- **`buscar(id_empleado)`** — restituisce il dipendente o `None`.
- **`salario_medio()`** — media degli stipendi (0 se non ci sono dipendenti).
- **`el_mas_pagado()`** — restituisce il dipendente con lo stipendio più alto.
- **`listar()`** — stampa tutti i dipendenti ordinati per cognome.
- **`__len__`** — numero di dipendenti.
- **`__contains__`** — permette `empleado in departamento`.
- **`__str__`** — `"Departamento Desarrollo (3 empleados, media: 2.766,67 €)"`.

class Departamento:
    def __init__(self, nombre):
        self.nombre     = nombre
        self._empleados = []

    def añadir(self, empleado):
        if not isinstance(empleado, Empleado):
            raise TypeError(f"Atteso un Empleado, non {type(empleado).__name__}")
        if empleado in self._empleados:
            print(f"Avviso: {empleado.nombre_completo()} è già nel dipartimento")
            return
        self._empleados.append(empleado)

    def eliminar(self, id_empleado):
        empleado = self.buscar(id_empleado)
        if empleado is None:
            raise ValueError(f"Non esiste nessun dipendente con id '{id_empleado}'")
        self._empleados.remove(empleado)
        return empleado

    def buscar(self, id_empleado):
        for emp in self._empleados:
            if emp.id == id_empleado:
                return emp
        return None

    def salario_medio(self):
        if not self._empleados:
            return 0.0
        return sum(e.salario for e in self._empleados) / len(self._empleados)

    def el_mas_pagado(self):
        if not self._empleados:
            return None
        return max(self._empleados)   # usa __lt__

    def listar(self):
        if not self._empleados:
            print(f"Il dipartimento {self.nombre} non ha dipendenti.")
            return
        print(f"\n{'─'*55}")
        print(f"  {self.nombre.upper()} — {len(self._empleados)} dipendente/i")
        print(f"{'─'*55}")
        for emp in sorted(self._empleados, key=lambda e: e.apellido):
            print(f"  {emp}")
        print(f"{'─'*55}")
        print(f"  Stipendio medio: {self.salario_medio():,.2f} €")
        mas_pagado = self.el_mas_pagado()
        print(f"  Più pagato/a:    {mas_pagado.nombre_completo()} ({mas_pagado.salario:,.2f} €)")

    # ─── Dunder ───────────────────────────────────────────
    def __len__(self):
        return len(self._empleados)

    def __contains__(self, empleado):
        return empleado in self._empleados

    def __str__(self):
        media = self.salario_medio()
        return (
            f"Departamento {self.nombre} "
            f"({len(self)} dipendente/i, media: {media:,.2f} €)"
        )


# --- Prova ---
Empleado._contador_ids = 0

dev = Departamento("Desarrollo")
mkt = Departamento("Marketing")

e1 = Empleado("Ana",   "García",  "Desarrollo", 2500)
e2 = Empleado("Luis",  "Pérez",   "Marketing",  3100)
e3 = Empleado("Marta", "López",   "Desarrollo", 2800)
e4 = Empleado("Eva",   "Martín",  "Desarrollo", 2700)

dev.añadir(e1); dev.añadir(e3); dev.añadir(e4)
mkt.añadir(e2)

dev.listar()

print(f"\n{dev}")
print(f"e1 è in dev? {e1 in dev}")
print(f"e2 è in dev? {e2 in dev}")

dev.eliminar(e3.id)
print(f"\nDopo aver eliminato {e3.nombre_completo()}: {dev}")
```

**Punti chiave:**
- `eliminar` usa `buscar` internamente — evita di duplicare la logica di ricerca.
- `max(self._empleados)` funziona grazie a `__lt__` definito in `Empleado`.
- `sorted(..., key=lambda e: e.apellido)` ordina alfabeticamente senza modificare la lista interna.
- `__contains__` delega a `list.__contains__`, che a sua volta usa `__eq__` di `Empleado` — la catena di dunder funziona da sola.

import json
from pathlib import Path

class Empresa:
    def __init__(self, nombre):
        self.nombre        = nombre
        self._departamentos = {}   # {nombre_dept: Departamento}

    # ─── Gestione dei dipendenti ─────────────────────────────
    def contratar(self, empleado, nombre_departamento):
        if nombre_departamento not in self._departamentos:
            self._departamentos[nombre_departamento] = Departamento(nombre_departamento)
        self._departamentos[nombre_departamento].añadir(empleado)
        print(f"✅ {empleado.nombre_completo()} assunto/a in {nombre_departamento}")

    def transferir(self, id_empleado, nuevo_departamento):
        empleado, dept_origen = self.buscar_global(id_empleado)
        if empleado is None:
            raise ValueError(f"Non esiste nessun dipendente con id '{id_empleado}'")
        dept_origen.eliminar(id_empleado)
        self.contratar(empleado, nuevo_departamento)
        print(f"🔄 {empleado.nombre_completo()} trasferito/a da {dept_origen.nombre} a {nuevo_departamento}")

    def despedir(self, id_empleado):
        empleado, dept = self.buscar_global(id_empleado)
        if empleado is None:
            raise ValueError(f"Non esiste nessun dipendente con id '{id_empleado}'")
        dept.eliminar(id_empleado)
        print(f"❌ {empleado.nombre_completo()} licenziato/a da {dept.nombre}")
        return empleado

    def buscar_global(self, id_empleado):
        for dept in self._departamentos.values():
            emp = dept.buscar(id_empleado)
            if emp is not None:
                return emp, dept
        return None, None

    # ─── Report ─────────────────────────────────────────
    def todos_los_empleados(self):
        return [emp for dept in self._departamentos.values() for emp in dept._empleados]

    def resumen(self):
        todos   = self.todos_los_empleados()
        n_total = len(todos)
        print(f"\n{'═'*60}")
        print(f"  {self.nombre.upper()}")
        print(f"  {n_total} dipendente/i in {len(self._departamentos)} dipartimento/i")
        print(f"{'═'*60}")

        for dept in self._departamentos.values():
            dept.listar()

        if todos:
            mas_pagado = max(todos)   # usa __lt__
            media      = sum(e.salario for e in todos) / n_total
            print(f"\n  ► Stipendio medio azienda: {media:,.2f} €")
            print(f"  ► Più pagato/a:            {mas_pagado.nombre_completo()} ({mas_pagado.salario:,.2f} €)")
        print(f"{'═'*60}")

    def exportar_json(self, ruta):
        datos = {
            "empresa": self.nombre,
            "departamentos": {
                nombre: [emp.info() for emp in dept._empleados]
                for nombre, dept in self._departamentos.items()
            }
        }
        Path(ruta).write_text(
            json.dumps(datos, indent=2, ensure_ascii=False),
            encoding="utf-8"
        )
        print(f"\n📁 Dati esportati in {ruta}")

    def __str__(self):
        n = len(self.todos_los_empleados())
        return f"{self.nombre} ({len(self._departamentos)} dip., {n} dipendenti)"

empleados = [

    Empleado("Marta",   "López",     "Desarrollo", 2800),
    Empleado("Carlos",  "Ruiz",      "Ventas",     2900),
    Empleado("Eva",     "Martín",    "Marketing",  2600),
    Empleado("Jorge",   "Fernández", "Desarrollo", 3400),
]

asignaciones = [
    (empleados[0], "Desarrollo"),
    (empleados[1], "Marketing"),
    (empleados[2], "Desarrollo"),
    (empleados[3], "Ventas"),
    (empleados[4], "Marketing"),
    (empleados[5], "Desarrollo"),
]

print("── Assunzioni ──")
for emp, dept in asignaciones:
    empresa.contratar(emp, dept)

print("\n── Trasferimento ──")
empresa.transferir(empleados[4].id, "Desarrollo")

print("\n── Licenziamento ──")
empresa.despedir(empleados[3].id)

empresa.resumen()
empresa.exportar_json("empresa.json")
```

**Punti chiave:**
- `transferir` riutilizza `buscar_global`, `eliminar` e `contratar` — senza duplicare la logica.

### Anatomia di una classe

```python
class MiaClasse:
    attributo_classe = valore             # condiviso

    def __init__(self, param):           # costruttore
        self.attributo = param           # di istanza
        self._privato = None             # convenzione: privato

    @property
    def privato(self):                   # getter
        return self._privato

    @privato.setter
    def privato(self, valore):           # setter con validazione
        self._privato = valore

    def metodo(self):                    # di istanza
        return self.attributo

    @classmethod
    def da_dict(cls, dati):              # costruttore alternativo
        return cls(dati["param"])

    @staticmethod
    def e_valido(valore):                # utilità senza self né cls
        return valore > 0

    def __str__(self):                   # per print() e str()
        return f"MiaClasse({self.attributo})"

    def __eq__(self, altro):             # per ==
        return self.attributo == altro.attributo
```

### Tabella di riferimento rapida

| Concetto | Sintassi | Uso |
|----------|---------|-----|
| Classe | `class Nome:` | Modello |
| Istanza | `obj = Nome(args)` | Copia concreta |
| Costruttore | `def __init__(self, ...)` | Inizializzare gli attributi |
| Attributo istanza | `self.x = valore` | Dati dell'oggetto |
| Attributo classe | `x = valore` (fuori dai metodi) | Dati condivisi |
| Metodo istanza | `def f(self)` | Lavora con l'oggetto |
| Metodo di classe | `@classmethod def f(cls)` | Lavora con la classe |
| Metodo statico | `@staticmethod def f()` | Utilità senza stato |
| Proprietà | `@property def x(self)` | Getter con logica |
| Setter | `@x.setter def x(self, v)` | Validare l'assegnazione |
