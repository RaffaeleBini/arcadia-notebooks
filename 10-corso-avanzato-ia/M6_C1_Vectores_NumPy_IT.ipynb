## 3. Cos'è un vettore?

```
Un vettore è una lista ordinata di numeri.

In 2D:  v = [3, 4]   → punto sul piano, o freccia dall'origine
In 3D:  v = [1, 2, 3] → punto nello spazio
Nell'IA:  v = [0.2, -0.5, 0.8, ..., 0.1]  → 768 dimensioni (BERT)

Ogni numero è una COORDINATA o COMPONENTE del vettore.

      ↑  y
   4  │      ●  (3, 4)
   3  │    ↗
   2  │  ↗
   1  │↗
      └──────→ x
         1  2  3
```

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Sinistra: vettori in 2D
ax = axes[0]
vectores = {
    "v = [3, 4]": ([3, 4], "blue"),
    "u = [-2, 3]": ([-2, 3], "red"),
    "w = [4, -1]": ([4, -1], "green"),
}
for nombre, (vec, color) in vectores.items():
    ax.annotate("", xy=(vec[0], vec[1]), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, lw=2))
    ax.text(vec[0]*0.5, vec[1]*0.5, nombre, color=color, fontsize=9)

ax.set_xlim(-3, 5); ax.set_ylim(-2, 5)
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.grid(True, alpha=0.3); ax.set_title("Vettori in 2D")
ax.set_xlabel("x"); ax.set_ylabel("y")

# Destra: somma di vettori
ax2 = axes[1]
v = np.array([3, 1])
u = np.array([1, 2])
suma = v + u

ax2.annotate("", xy=v, xytext=(0,0), arrowprops=dict(arrowstyle="->", color="blue", lw=2))
ax2.annotate("", xy=suma, xytext=v, arrowprops=dict(arrowstyle="->", color="red", lw=2, linestyle="dashed"))
ax2.annotate("", xy=suma, xytext=(0,0), arrowprops=dict(arrowstyle="->", color="purple", lw=2.5))
ax2.text(v[0]/2, v[1]/2-0.2, "v=[3,1]", color="blue", fontsize=9)
ax2.text(v[0]+0.1, v[1]+0.5, "u=[1,2]", color="red", fontsize=9)
ax2.text(suma[0]/2-0.5, suma[1]/2+0.2, f"v+u={list(suma)}", color="purple", fontsize=9, fontweight="bold")

ax2.set_xlim(-1, 6); ax2.set_ylim(-1, 4)
ax2.axhline(0, color='gray', lw=0.5); ax2.axvline(0, color='gray', lw=0.5)
ax2.grid(True, alpha=0.3); ax2.set_title("Somma di vettori")
ax2.set_xlabel("x"); ax2.set_ylabel("y")

plt.tight_layout()
plt.show()
````

## 4. Magnitudine e direzione

```
MAGNITUDINE (modulo, norma) = lunghezza del vettore

  Per v = [3, 4]:
  ‖v‖ = √(3² + 4²) = √(9 + 16) = √25 = 5

  In generale:
  ‖v‖ = √(v₁² + v₂² + ... + vₙ²)

DIREZIONE = vettore unitario (magnitudine = 1)

  v̂ = v / ‖v‖

  Per v = [3, 4]:
  v̂ = [3/5, 4/5] = [0.6, 0.8]
  ‖v̂‖ = √(0.6² + 0.8²) = √(0.36 + 0.64) = 1  ✓

A cosa serve il vettore unitario?
→ Separa la lunghezza (quanto) dalla direzione (verso dove)
→ Nell'IA: la similarità coseno usa vettori unitari
```

````python
v = np.array([3, 4])

# Magnitudine manuale
magnitud_manual = np.sqrt(np.sum(v ** 2))

# Magnitudine con NumPy
magnitud_numpy = np.linalg.norm(v)

print(f"Vettore:            {v}")
print(f"Magnitudine (manuale): {magnitud_manual}")
print(f"Magnitudine (numpy):   {magnitud_numpy}")

# Vettore unitario (normalizzare)
v_unitario = v / np.linalg.norm(v)
print(f"\nVettore unitario: {v_unitario}")
print(f"‖v̂‖ = {np.linalg.norm(v_unitario):.6f}  (deve essere 1)")

# Diversi vettori con magnitudine diversa, stessa direzione
print("\n── Stessa direzione, magnitudine diversa ──")
for escalar in [0.5, 1, 2, 5]:
    w = v * escalar
    w_u = w / np.linalg.norm(w)
    print(f"  {escalar}×v = {w}  → ‖w‖={np.linalg.norm(w):.1f}  → v̂={np.round(w_u, 4)}")

print("→ Il vettore unitario è sempre lo stesso anche se la magnitudine cambia")
````

## 6. Prodotto scalare (dot product)

```
v · u = v₁×u₁ + v₂×u₂ + ... + vₙ×uₙ

Inoltre: v · u = ‖v‖ × ‖u‖ × cos(θ)

Dove θ è l'angolo tra i vettori.

Interpretazione:
  v · u > 0  → i vettori puntano in direzione simile   (θ < 90°)
  v · u = 0  → vettori perpendicolari                  (θ = 90°)
  v · u < 0  → i vettori puntano in direzione opposta  (θ > 90°)

Nell'IA: si usa costantemente nelle reti neurali (ogni neurone
calcola il prodotto scalare tra i suoi pesi e l'input).
```

````python
v = np.array([1, 2, 3])
u = np.array([4, 5, 6])

# Tre modi per calcolare il prodotto scalare
dot_manual = sum(vi * ui for vi, ui in zip(v, u))
dot_numpy1 = np.dot(v, u)
dot_numpy2 = v @ u          # operatore @ (Python 3.5+)

print(f"v = {v}")
print(f"u = {u}")
print(f"v · u (manuale) = {1*4} + {2*5} + {3*6} = {dot_manual}")
print(f"v · u (np.dot)  = {dot_numpy1}")
print(f"v · u (op. @)   = {dot_numpy2}")

# Verifica con la formula dell'angolo
cos_theta = dot_numpy1 / (np.linalg.norm(v) * np.linalg.norm(u))
theta_rad = np.arccos(cos_theta)
theta_grados = np.degrees(theta_rad)
print(f"\nAngolo tra v e u: {theta_grados:.2f}°")

# Casi speciali
print("\n── Casi speciali ──")
a = np.array([1, 0])
b_paralelo    = np.array([3, 0])  # stessa direzione
b_perpendic   = np.array([0, 2])  # perpendicolare
b_opuesto     = np.array([-2, 0]) # direzione opposta

for nombre, b in [("parallelo", b_paralelo), ("perpendicolare", b_perpendic), ("opposto", b_opuesto)]:
    print(f"  a={a} · b_{nombre}={b}: dot={np.dot(a,b)}, angolo={np.degrees(np.arccos(np.clip(np.dot(a/np.linalg.norm(a), b/np.linalg.norm(b)), -1, 1))):.0f}°")
````

## 7. Similarità coseno — applicazione reale nell'IA

```
similarita_coseno(v, u) = (v · u) / (‖v‖ × ‖u‖)

Risultato tra -1 e 1:
   1   → vettori identici in direzione
   0   → vettori perpendicolari (nessuna relazione)
  -1   → vettori opposti

Vantaggio rispetto alla distanza euclidea:
→ La magnitudine non conta, solo la direzione
→ "gatto" × 1000 è simile a "cane" tanto quanto "gatto" × 1

Uso nell'IA:
→ Ricerca semantica (trovare testi simili)
→ Sistemi di raccomandazione
→ Rilevamento di duplicati
```

````python
def similitud_coseno(v, u):
    return np.dot(v, u) / (np.linalg.norm(v) * np.linalg.norm(u))


# Simuliamo embedding di parole (in realtà sono vettori di 768+ dimensioni)
# Usiamo 4 dimensioni perché sia comprensibile
# Dimensioni: [animale, domestico, meccanico, veloce]
embeddings = {
    "gatto":     np.array([0.9,  0.8, -0.1,  0.3]),
    "cane":      np.array([0.8,  0.9, -0.1,  0.5]),
    "tigre":     np.array([0.9, -0.3, -0.1,  0.7]),
    "auto":      np.array([-0.1, -0.2,  0.9,  0.8]),
    "bicicletta":np.array([-0.1, -0.1,  0.7,  0.4]),
    "aereo":     np.array([-0.1, -0.2,  0.8,  0.9]),
}

print("Similarità coseno tra parole:")
print(f"{'Coppia':<25} {'Similarità':>10} {'Interpretazione'}")
print("-" * 60)

pares = [
    ("gatto",     "cane"),
    ("gatto",     "tigre"),
    ("gatto",     "auto"),
    ("auto",      "bicicletta"),
    ("auto",      "aereo"),
    ("gatto",     "gatto"),
]
for p1, p2 in pares:
    sim = similitud_coseno(embeddings[p1], embeddings[p2])
    if sim > 0.9:
        interp = "molto simili"
    elif sim > 0.6:
        interp = "correlate"
    elif sim > 0.2:
        interp = "un po' correlate"
    else:
        interp = "nessuna relazione"
    print(f"  {p1} ↔ {p2:<12} {sim:>8.4f}   {interp}")

# Visualizzazione della mappa di similarità
palabras = list(embeddings.keys())
n = len(palabras)
matriz_sim = np.zeros((n, n))
for i, p1 in enumerate(palabras):
    for j, p2 in enumerate(palabras):
        matriz_sim[i, j] = similitud_coseno(embeddings[p1], embeddings[p2])

fig, ax = plt.subplots(figsize=(7, 5))
im = ax.imshow(matriz_sim, cmap="RdYlGn", vmin=-1, vmax=1)
ax.set_xticks(range(n)); ax.set_yticks(range(n))
ax.set_xticklabels(palabras, rotation=45, ha="right")
ax.set_yticklabels(palabras)
for i in range(n):
    for j in range(n):
        ax.text(j, i, f"{matriz_sim[i,j]:.2f}", ha="center", va="center", fontsize=8)
plt.colorbar(im, ax=ax, label="Similarità coseno")
ax.set_title("Mappa di similarità tra parole")
plt.tight_layout()
plt.show()
````

1. Quale coppia di clienti è più simile secondo la distanza euclidea?

<summary>💡 Vedi soluzione — Esercizio 3</summary>

```python
import numpy as np
from itertools import combinations

clientes = {
    "C1": np.array([35, 45000, 8,  3, 4.2]),
    "C2": np.array([37, 48000, 9,  4, 4.5]),
    "C3": np.array([28, 28000, 3,  1, 3.1]),
    "C4": np.array([55, 80000, 15, 8, 4.8]),
    "C5": np.array([36, 46000, 8,  3, 4.3]),
}

nombres = list(clientes.keys())
vectores = list(clientes.values())

# 1. Distanza euclidea
print("Distanze euclidee:")
min_dist, par_dist = float('inf'), None
for n1, n2 in combinations(nombres, 2):
    dist = np.linalg.norm(clientes[n1] - clientes[n2])
    print(f"  {n1} ↔ {n2}: {dist:.2f}")
    if dist < min_dist:
        min_dist, par_dist = dist, (n1, n2)
print(f"\nCoppia più simile (euclidea): {par_dist} ({min_dist:.2f})")

# 2. Similarità coseno
print("\nSimilarità coseno:")
max_sim, par_cos = -1, None
for n1, n2 in combinations(nombres, 2):
    v1, v2 = clientes[n1], clientes[n2]
    sim = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
    print(f"  {n1} ↔ {n2}: {sim:.6f}")
    if sim > max_sim:
        max_sim, par_cos = sim, (n1, n2)
print(f"\nCoppia più simile (coseno): {par_cos} ({max_sim:.6f})")

# 3. Normalizzare (z-score)
matriz = np.vstack(vectores)
media   = matriz.mean(axis=0)
std     = matriz.std(axis=0)
norm    = (matriz - media) / std
clientes_norm = {n: norm[i] for i, n in enumerate(nombres)}

print("\nDistanze con normalizzazione z-score:")
min_dist_n, par_dist_n = float('inf'), None
for n1, n2 in combinations(nombres, 2):
    dist = np.linalg.norm(clientes_norm[n1] - clientes_norm[n2])
    print(f"  {n1} ↔ {n2}: {dist:.4f}")
    if dist < min_dist_n:
        min_dist_n, par_dist_n = dist, (n1, n2)
print(f"\nCoppia più simile (normalizzata): {par_dist_n}")
```

```python
import numpy as np

def similitud_coseno(v, u):
    v, u = np.array(v), np.array(u)
    return np.dot(v, u) / (np.linalg.norm(v) * np.linalg.norm(u))

def buscar(consulta, documentos, top_n=None):
    similitudes = [
        (titulo, similitud_coseno(consulta, vec))
        for titulo, vec in documentos.items()
    ]
    similitudes.sort(key=lambda x: x[1], reverse=True)
```

```python
import numpy as np

peliculas = {
    "Interstellar":    [0.3, 0.1, 0.6, 0.0, 0.9, 0.1],
    "The Dark Knight": [0.9, 0.1, 0.7, 0.2, 0.3, 0.0],
    "Toy Story":       [0.2, 0.9, 0.4, 0.0, 0.3, 0.2],
    "The Conjuring":   [0.1, 0.0, 0.4, 0.9, 0.1, 0.0],
    "La La Land":      [0.1, 0.3, 0.8, 0.0, 0.0, 0.9],
    "Avengers":        [0.9, 0.4, 0.3, 0.1, 0.8, 0.1],
    "Hereditary":      [0.0, 0.0, 0.7, 0.9, 0.2, 0.0],
    "Her":             [0.0, 0.2, 0.8, 0.0, 0.7, 0.6],
    "Superbad":        [0.1, 0.9, 0.3, 0.0, 0.0, 0.2],
    "Arrival":         [0.1, 0.0, 0.8, 0.1, 0.9, 0.3],
}
peliculas = {k: np.array(v) for k, v in peliculas.items()}
GENEROS = ["azione", "commedia", "drammatico", "horror", "sci-fi", "romantico"]

def similitud_coseno(v, u):
    return np.dot(v, u) / (np.linalg.norm(v) * np.linalg.norm(u))

def recomendar(pelicula_vista, n=3):
    base = peliculas[pelicula_vista]
    similitudes = [
        (titulo, similitud_coseno(base, vec))
        for titulo, vec in peliculas.items()
        if titulo != pelicula_vista
    ]
    return sorted(similitudes, key=lambda x: x[1], reverse=True)[:n]

def perfil_usuario(peliculas_vistas):
    vectores = [peliculas[p] for p in peliculas_vistas]
    return np.mean(vectores, axis=0)

def recomendar_para_usuario(peliculas_vistas, n=3):
    perfil = perfil_usuario(peliculas_vistas)
    print(f"Profilo dell'utente (generi più affini):")
    for g, v in sorted(zip(GENEROS, perfil), key=lambda x: -x[1]):
        print(f"  {g}: {v:.2f}")
    candidatas = {t: v for t, v in peliculas.items() if t not in peliculas_vistas}
    similitudes = [(t, similitud_coseno(perfil, v)) for t, v in candidatas.items()]
    return sorted(similitudes, key=lambda x: x[1], reverse=True)[:n]

print("Se ti è piaciuto 'Interstellar', ti raccomandiamo:")
for titulo, sim in recomendar("Interstellar"):
    print(f"  {sim:.4f}  {titulo}")

vistas = ["Interstellar", "Arrival", "Her"]
print(f"\nL'utente ha visto: {vistas}")
recomendaciones = recomendar_para_usuario(vistas)
print("\nRaccomandazioni:")
for titulo, sim in recomendaciones:
    print(f"  {sim:.4f}  {titulo}")
```
