## 1. ¿Por qué necesitamos bucles?

Imagina que quieres imprimir los números del 1 al 5. Sin bucles:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

Funciona. Pero, ¿y si son del 1 al 1 000? ¿O del 1 al 1 000 000?

Con un bucle, cualquiera de esos casos es exactamente el mismo código — solo cambia el número final.

Hay **dos tipos de bucle** en Python:

| Bucle | Cuándo usarlo |
|-------|---------------|
| `while` | Cuando no sabes cuántas veces vas a repetir — repites **mientras** se cumpla una condición |
| `for` | Cuando sabes cuántas veces vas a repetir — recorres una secuencia de valores |

## 2. `while` — Repetir mientras se cumpla una condición

### Sintaxis

```python
while condicion:
    # bloque que se repite mientras condicion sea True
```

Python evalúa la condición **antes de cada iteración**. Si es `True`, ejecuta el bloque y vuelve a comprobar. Si es `False`, sale del bucle.

```
┌─────────────────────────────┐
│   ¿condicion es True?       │
│        │                    │
│    Sí  │        No          │
│        ▼          ▼         │
│   ejecuta      sale del     │
│   bloque       bucle        │
│       │                     │
│       └──────── vuelve ─────┘
```

> ⚠️ **Bucle infinito:** si la condición nunca se vuelve `False`, el bucle no termina. Siempre asegúrate de que algo dentro del bucle modifique la condición.

````python
# Contar del 1 al 5
contador = 1

while contador <= 5:
    print(contador)
    contador += 1     # sin esta línea → bucle infinito

print("Bucle terminado")
````

````python
# Pedir un número positivo hasta que el usuario lo introduzca correctamente
numero = -1   # valor inicial que fuerza entrar al bucle

while numero <= 0:
    numero = int(input("Introduce un número positivo: "))
    if numero <= 0:
        print("Ese número no es positivo, inténtalo de nuevo.")

print(f"Perfecto, has introducido: {numero}")
````

````python
# Cuenta atrás
cuenta = 5

while cuenta > 0:
    print(f"{cuenta}...")
    cuenta -= 1

print("¡Despegue! 🚀")
````

### `while` con `else`

Python permite añadir un bloque `else` a un `while`. Se ejecuta **cuando el bucle termina de forma natural** (la condición se vuelve `False`), pero **no** si se sale con `break`.

````python
intentos = 3

while intentos > 0:
    clave = input("Introduce la clave: ")
    if clave == "python":
        print("Acceso concedido ✅")
        break               # sale del bucle sin pasar por el else
    intentos -= 1
    print(f"Clave incorrecta. Intentos restantes: {intentos}")
else:
    # solo llega aquí si el while terminó sin break (se agotaron los intentos)
    print("Cuenta bloqueada ❌")
````

## 3. `for` — Recorrer una secuencia

El bucle `for` recorre uno a uno los elementos de una **secuencia** (una lista, un rango, un string…) y ejecuta el bloque para cada elemento.

### Sintaxis

```python
for variable in secuencia:
    # bloque que se ejecuta una vez por cada elemento
    # 'variable' toma el valor del elemento actual
```

En cada iteración, `variable` **toma automáticamente** el valor del siguiente elemento de la secuencia.

```
secuencia = [10, 20, 30]

iteración 1: variable = 10  → ejecuta bloque
iteración 2: variable = 20  → ejecuta bloque
iteración 3: variable = 30  → ejecuta bloque
             (fin de secuencia → sale)
```

````python
# Recorrer una lista de elementos
frutas = ["manzana", "naranja", "plátano", "uva"]

for fruta in frutas:
    print(f"Fruta: {fruta}")
````

````python
# Un string también es una secuencia: se recorre letra a letra
palabra = "Python"

for letra in palabra:
    print(letra)
````

### 3.1 `range()` — Generar secuencias de números

`range()` genera una secuencia de números sin ocupar memoria. Es la forma más habitual de controlar cuántas veces se repite un `for`.

| Forma | Genera | Ejemplo |
|-------|--------|---------|
| `range(n)` | 0, 1, 2, … n-1 | `range(5)` → 0 1 2 3 4 |
| `range(inicio, fin)` | inicio, …, fin-1 | `range(2, 6)` → 2 3 4 5 |
| `range(inicio, fin, paso)` | de inicio a fin con saltos | `range(0, 10, 2)` → 0 2 4 6 8 |

> El valor `fin` **nunca se incluye** — es como los índices de los strings.

````python
# range(n): del 0 al n-1
print("range(5):")
for i in range(5):
    print(i, end=" ")   # end=" " imprime en la misma línea
print()

# range(inicio, fin)
print("range(1, 6):")
for i in range(1, 6):
    print(i, end=" ")
print()

# range(inicio, fin, paso)
print("range(0, 20, 3):")
for i in range(0, 20, 3):
    print(i, end=" ")
print()

# Paso negativo: cuenta hacia atrás
print("range(10, 0, -2):")
for i in range(10, 0, -2):
    print(i, end=" ")
print()
````

````python
# Tabla de multiplicar del 7
numero = 7

print(f"Tabla del {numero}:")
for i in range(1, 11):
    print(f"  {numero} × {i:2} = {numero * i:3}")
````

### 3.2 `enumerate()` — Índice y valor a la vez

Cuando necesitas tanto la **posición** como el **valor** de cada elemento, usa `enumerate()`. Evita el patrón `for i in range(len(lista))` que es menos legible.

````python
asignaturas = ["Matemáticas", "Lengua", "Historia", "Física"]

# Sin enumerate (menos idiomático)
for i in range(len(asignaturas)):
    print(f"{i + 1}. {asignaturas[i]}")

print("---")

# Con enumerate (más claro)
for posicion, asignatura in enumerate(asignaturas, start=1):
    print(f"{posicion}. {asignatura}")
````

### 3.3 Bucles `for` anidados

Un `for` dentro de otro `for`. Por cada iteración del bucle exterior, el bucle interior completa **todas** sus iteraciones.

````python
# Tabla de multiplicar completa (del 1 al 5)
for i in range(1, 6):           # bucle exterior: filas
    for j in range(1, 6):       # bucle interior: columnas
        print(f"{i*j:3}", end=" ")
    print()                     # salto de línea al terminar cada fila
````

## 4. `break` — Salir del bucle antes de tiempo

`break` **interrumpe el bucle inmediatamente** y salta a la primera instrucción después del bucle, ignorando las iteraciones restantes.

```
for / while ...:
    ...
    if condicion_de_salida:
        break           ← sale del bucle
    ...
código después del bucle   ← continúa aquí
```

Se usa cuando encontramos lo que buscábamos y no tiene sentido seguir iterando.

````python
# Buscar el primer número divisible por 7 entre 1 y 100
for numero in range(1, 101):
    if numero % 7 == 0:
        print(f"Primer múltiplo de 7: {numero}")
        break   # ya lo encontramos, no necesitamos seguir
    print(f"{numero} no es múltiplo de 7")
````

````python
# Buscar una letra en un texto
texto  = "programación"
buscar = "a"
encontrado = False

for i, letra in enumerate(texto):
    if letra == buscar:
        print(f"'{buscar}' encontrada en la posición {i}")
        encontrado = True
        break

if not encontrado:
    print(f"'{buscar}' no está en el texto")
````

## 5. `continue` — Saltar al siguiente ciclo

`continue` **salta el resto del bloque actual** y pasa directamente a la siguiente iteración del bucle. No sale del bucle — solo se salta lo que queda de esa iteración.

```
for / while ...:
    ...
    if condicion_de_salto:
        continue    ← salta al siguiente ciclo
    ...             ← esto NO se ejecuta si continue fue activado
```

Se usa para **filtrar** elementos que no queremos procesar.

````python
# Imprimir solo los números impares del 1 al 10
for n in range(1, 11):
    if n % 2 == 0:
        continue    # salta los pares
    print(n, end=" ")
print()
````

````python
# Procesar una lista de notas, ignorando las inválidas
notas = [8.5, -1, 6.0, 11, 7.3, 4.5, 9.1, -3, 5.0]

notas_validas = 0
suma          = 0

for nota in notas:
    if nota < 0 or nota > 10:
        print(f"  Nota inválida ignorada: {nota}")
        continue    # salta esta nota
    notas_validas += 1
    suma          += nota

if notas_validas > 0:
    media = suma / notas_validas
    print(f"\nNotas válidas: {notas_validas}")
    print(f"Media: {media:.2f}")
````

### Diferencia visual entre `break` y `continue`

```
Lista: [1, 2, 3, 4, 5]
Condición: número == 3

Con BREAK:
  1 ✓  2 ✓  3 → PARA  (4 y 5 no se procesan)

Con CONTINUE:
  1 ✓  2 ✓  3 → SALTA  4 ✓  5 ✓  (3 no se procesa, pero 4 y 5 sí)
```

````python
numeros = [1, 2, 3, 4, 5]

print("Con break:")
for n in numeros:
    if n == 3:
        break
    print(n, end=" ")
print()

print("Con continue:")
for n in numeros:
    if n == 3:
        continue
    print(n, end=" ")
print()
````

## 6. Patrones comunes con bucles

Hay tres patrones que aparecen constantemente — merece la pena reconocerlos:

````python
# PATRÓN 1: CONTADOR
# Cuenta cuántos elementos cumplen una condición

notas = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
aprobados = 0

for nota in notas:
    if nota >= 5:
        aprobados += 1

print(f"Aprobados: {aprobados} de {len(notas)}")
````

````python
# PATRÓN 2: ACUMULADOR
# Acumula un valor (suma, producto, construcción de string…)

notas = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
suma  = 0

for nota in notas:
    suma += nota

media = suma / len(notas)
print(f"Suma: {suma}")
print(f"Media: {media:.1f}")

# Python tiene funciones integradas para esto:
print(f"Con sum(): {sum(notas)}")
print(f"Con max(): {max(notas)}")
print(f"Con min(): {min(notas)}")
````

````python
# PATRÓN 3: BÚSQUEDA
# Encuentra el primero que cumple una condición

notas = [4, 7, 3, 9, 5, 2, 8, 6, 1, 10]
primera_nota_alta = None

for nota in notas:
    if nota >= 9:
        primera_nota_alta = nota
        break

if primera_nota_alta is not None:
    print(f"Primera nota >= 9: {primera_nota_alta}")
else:
    print("No hay notas >= 9")
````

### ¿Cuándo usar `while` y cuándo `for`?

| Situación | Usa |
|-----------|-----|
| Recorrer una lista, string o rango | `for` |
| Repetir exactamente N veces | `for` con `range(N)` |
| Repetir hasta que el usuario introduzca un valor válido | `while` |
| Repetir hasta que ocurra algo (sin saber cuándo) | `while` |
| Bucle infinito controlado con `break` | `while True:` |

El patrón `while True:` con `break` es muy habitual para **menús** o **bucles de juego**:

````python
# Patrón while True: el bucle solo sale cuando el usuario elige salir
while True:
    opcion = input("\nMenú: (1) Saludar  (2) Despedirse  (3) Salir → ")

    if opcion == "1":
        print("¡Hola!")
    elif opcion == "2":
        print("¡Hasta luego!")
    elif opcion == "3":
        print("Saliendo...")
        break
    else:
        print("Opción no válida")
````

**Ejemplo de salida esperada:**
```
Adivina el número (entre 1 y 100): 50
Demasiado alto.
Adivina el número (entre 1 y 100): 25
Demasiado bajo.
Adivina el número (entre 1 y 100): 42
```

````python
import random
````

- `{intento:3}` alinea los números a la derecha en un campo de 3 caracteres — hace el historial más legible.

### Ejercicio 5 (reto) — Juego completo con múltiples partidas

**Enunciado:** Añade un bucle exterior que permita **jugar varias partidas**. Al terminar cada partida, pregunta si el jugador quiere volver a jugar. Al salir, muestra las estadísticas globales:

- Partidas jugadas
- Partidas ganadas
- Partidas perdidas
- Porcentaje de victorias
- Mejor resultado (menos intentos en una victoria)

**Ejemplo de salida al final:**
```
¿Quieres jugar otra partida? (s/n): n

=== Estadísticas finales ===
Partidas jugadas:   3
Victorias:          2
Derrotas:           1
% de victorias:     66.67%
Mejor resultado:    3 intentos
```

````python
# ✏️ TU CÓDIGO AQUÍ
import random
````

<details>
<summary>💡 Ver solución — Ejercicio 5</summary>

```python
import random

# Estadísticas globales
partidas_jugadas = 0
victorias        = 0
mejor_resultado  = None   # None hasta que haya al menos una victoria

while True:   # bucle de partidas
    secreto      = random.randint(1, 100)
    max_intentos = 6
    intentos     = 0
    ganado       = False

    print(f"\n=== Partida {partidas_jugadas + 1} ===")
    print("He pensado un número entre 1 y 100.")

    while intentos < max_intentos and not ganado:
        intento = int(input(f"Intento {intentos + 1}/{max_intentos} → "))

        if intento < 1 or intento > 100:
            print("⚠️ Fuera de rango.")
            continue

        intentos += 1

        if intento > secreto:
            print("Demasiado alto.")
        elif intento < secreto:
            print("Demasiado bajo.")
        else:
            ganado = True

    # Resultado de la partida
    partidas_jugadas += 1

    if ganado:
        victorias += 1
        print(f"¡Correcto! Lo conseguiste en {intentos} intentos.")
        # Actualizar mejor resultado
        if mejor_resultado is None or intentos < mejor_resultado:
            mejor_resultado = intentos
    else:
        print(f"¡Se acabaron los intentos! El número era {secreto}.")

    # ¿Otra partida?
    otra = input("\n¿Quieres jugar otra partida? (s/n): ")
    if otra.lower() != "s":
        break

# Estadísticas finales
derrotas    = partidas_jugadas - victorias
porcentaje  = (victorias / partidas_jugadas) * 100 if partidas_jugadas > 0 else 0

print("\n=== Estadísticas finales ===")
print(f"Partidas jugadas:   {partidas_jugadas}")
print(f"Victorias:          {victorias}")
print(f"Derrotas:           {derrotas}")
print(f"% de victorias:     {porcentaje:.2f}%")

if mejor_resultado is not None:
    print(f"Mejor resultado:    {mejor_resultado} intentos")
else:
    print("Mejor resultado:    sin victorias")
```

**Puntos clave:**
- Dos bucles anidados: el exterior controla las partidas, el interior controla cada intento.
- `mejor_resultado = None` al inicio permite distinguir entre "todavía sin victorias" y "victoria con 0 intentos" (imposible, pero es buena práctica).
- La actualización del mejor resultado usa `is None or intentos < mejor_resultado` — la primera condición cubre la primera victoria.
- `porcentaje` usa el operador ternario para evitar división por cero si `partidas_jugadas == 0`.
</details>

## Resumen de la clase

### Los dos bucles

```python
# while — repite mientras la condición sea True
while condicion:
    bloque

# for — recorre una secuencia elemento a elemento
for variable in secuencia:
    bloque

# range — genera secuencias de números
range(n)              # 0 … n-1
range(inicio, fin)    # inicio … fin-1
range(inicio, fin, paso)
```

### Control de flujo dentro de bucles

```python
break      # sale del bucle inmediatamente
continue   # salta al siguiente ciclo
```

### Patrones fundamentales

```python
# Contador
cuenta = 0
for x in secuencia:
    if condicion(x):
        cuenta += 1

# Acumulador
total = 0
for x in secuencia:
    total += x

# Búsqueda
for x in secuencia:
    if es_lo_que_busco(x):
        resultado = x
        break

# Bucle de menú
while True:
    opcion = input("...")
    if opcion == "salir":
        break
```

### Funciones útiles

```python
enumerate(iterable, start=0)   # índice + valor
sum(iterable)                  # suma todos los elementos
max(iterable)                  # elemento máximo
min(iterable)                  # elemento mínimo
len(iterable)                  # número de elementos
random.randint(a, b)           # entero aleatorio entre a y b (ambos incluidos)
```
