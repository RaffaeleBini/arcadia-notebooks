```python
corpus_grande = [
    'el gato duerme en el sofá',
    'el perro juega en el jardín',
    'el gato y el perro son amigos',
    'a mi perro le encanta correr por el parque',
    'mi gato prefiere dormir todo el día'
]

# Tokenizador con vocabulario amplio: apenas afecta a este corpus pequeño
tokenizer_amplio = Tokenizer(num_words=100, oov_token="<OOV>")
tokenizer_amplio.fit_on_texts(corpus_grande)

# Tokenizador con vocabulario muy reducido: con num_words=5 solo se conservan los
# índices 1 a 4 al generar secuencias. El índice 1 es el token OOV, así que en la
# práctica quedan disponibles las 3 palabras más frecuentes; el resto pasa a ser OOV
tokenizer_reducido = Tokenizer(num_words=5, oov_token="<OOV>")
tokenizer_reducido.fit_on_texts(corpus_grande)

print("Vocabulario completo (word_index es igual en ambos, num_words solo afecta a texts_to_sequences):")
print(tokenizer_amplio.word_index)

# Secuencias con el tokenizador reducido: las palabras poco frecuentes se vuelven token 1 (OOV)
secuencias_reducidas = tokenizer_reducido.texts_to_sequences(corpus_grande)
print("Secuencias con num_words=5:", secuencias_reducidas)

total_oov = sum(token == 1 for secuencia in secuencias_reducidas for token in secuencia)
print("Tokens OOV encontrados:", total_oov)

# Sin truncar: se ve la longitud real de cada secuencia
sin_truncar = pad_sequences(secuencias_reducidas)
print("Sin maxlen (referencia):")
print(sin_truncar)

# Con maxlen=4 y truncating='post': las frases largas pierden sus últimas palabras
con_maxlen = pad_sequences(secuencias_reducidas, maxlen=4, truncating='post')
print("Con maxlen=4 y truncating='post':")
print(con_maxlen)
```

Con un vocabulario tan reducido (`num_words=5`), casi todas las palabras del corpus se convierten en el token OOV, y las secuencias resultantes se parecen mucho entre sí aunque las frases originales sean distintas: el modelo pierde información. Con `maxlen=4`, las frases más largas del corpus se recortan y pierden sus últimas palabras, que es justo lo que indica `truncating='post'`.
