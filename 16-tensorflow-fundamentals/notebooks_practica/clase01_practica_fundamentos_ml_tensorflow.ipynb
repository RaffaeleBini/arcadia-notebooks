## Ejercicio 2 — Clasifica un conjunto de datos sencillo

Hasta ahora entrenaste modelos de **regresión**: predicen un número (Y) a partir de otro número (X). Ahora vas a entrenar un modelo de **clasificación**: en vez de predecir un número, va a decidir a qué categoría (0 o 1) pertenece cada ejemplo.

El caso de uso: tienes puntos en dos dimensiones (x1, x2) y quieres que el modelo aprenda a distinguir si un punto está "por encima" o "por debajo" de la recta x1 + x2 = 0. Es decir:

- Si x1 + x2 > 0, la etiqueta es 1
- Si x1 + x2 <= 0, la etiqueta es 0

El andamiaje ya genera los datos de entrenamiento por ti con NumPy. Tu trabajo es completar el modelo.

Pistas:

- Sigue usando una capa `Dense`, pero esta vez con `activation='sigmoid'`: esa función convierte la salida en un valor entre 0 y 1, interpretable como una probabilidad
- La entrada ahora tiene dos valores (x1 y x2), así que `input_shape` debe ser `[2]`
- Para clasificación binaria, la función de pérdida adecuada es `'binary_crossentropy'`, no `'mean_squared_error'`
- Usa `optimizer='adam'` y añade `metrics=['accuracy']` en `compile` para ver la precisión durante el entrenamiento
- Entrena con menos épocas que en el ejercicio de regresión (con 100 es suficiente para este conjunto de datos tan sencillo)

````python
# Generamos 200 puntos aleatorios en dos dimensiones, entre -1 y 1
np.random.seed(42)
puntos = np.random.uniform(-1, 1, size=(200, 2))

# Etiqueta: 1 si x1 + x2 > 0, 0 en caso contrario
etiquetas = (puntos.sum(axis=1) > 0).astype(float)

# TODO: define el modelo con una capa Dense de una neurona,
# activation='sigmoid' e input_shape=[2]
modelo_clasificacion = None

# TODO: compila el modelo con optimizer='adam', loss='binary_crossentropy'
# y metrics=['accuracy']

# TODO: entrena el modelo con fit durante 100 épocas usando 'puntos' y 'etiquetas'

# TODO: prueba el modelo con un punto nuevo, por ejemplo [0.5, 0.5] (debería predecir cerca de 1)
# y con [-0.5, -0.5] (debería predecir cerca de 0)
````

<details>
<summary>💡 Ver solución</summary>

```python
np.random.seed(42)
puntos = np.random.uniform(-1, 1, size=(200, 2))
etiquetas = (puntos.sum(axis=1) > 0).astype(float)

# Una capa con una neurona y activación sigmoide, que recibe dos valores de entrada (x1, x2)
modelo_clasificacion = tf.keras.Sequential([
    keras.layers.Dense(units=1, activation='sigmoid', input_shape=[2])
])

# Para clasificación binaria usamos binary_crossentropy como función de pérdida
modelo_clasificacion.compile(optimizer='adam',
                              loss='binary_crossentropy',
                              metrics=['accuracy'])

# Entrenamos durante 100 épocas
modelo_clasificacion.fit(puntos, etiquetas, epochs=100, verbose=0)

# Probamos con dos puntos nuevos
print("Predicción para [0.5, 0.5]:", modelo_clasificacion.predict(np.array([[0.5, 0.5]])))
print("Predicción para [-0.5, -0.5]:", modelo_clasificacion.predict(np.array([[-0.5, -0.5]])))
```
</details>
