## Preparación común

Esta celda carga y normaliza Fashion MNIST, y la deja lista en formato `(28, 28, 1)` para usarla con capas `Conv2D`. Ejecútala antes de empezar cualquiera de los dos ejercicios.

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print('Versión de TensorFlow:', tf.__version__)

fashion_mnist = tf.keras.datasets.fashion_mnist
(training_images, training_labels), (test_images, test_labels) = fashion_mnist.load_data()

class_names = ['Camiseta', 'Pantalón', 'Jersey', 'Vestido', 'Abrigo',
               'Sandalia', 'Camisa', 'Zapatilla', 'Bolso', 'Botín']

# Normalizamos y añadimos el canal (28, 28) -> (28, 28, 1)
training_images = (training_images / 255.0).reshape(-1, 28, 28, 1)
test_images = (test_images / 255.0).reshape(-1, 28, 28, 1)

print('Forma de entrenamiento:', training_images.shape)
print('Forma de prueba:', test_images.shape)
````
