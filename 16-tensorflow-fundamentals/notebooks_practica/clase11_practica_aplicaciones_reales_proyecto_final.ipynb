```python
# Solución de referencia sobre Fashion MNIST (adapta los datos si usas otro dataset)
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Carga de datos
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()

# 2. Normalización de las entradas (los píxeles van de 0 a 255)
train_images = train_images / 255.0
test_images = test_images / 255.0

# 3. Modelo: una red densa sencilla es suficiente como referencia inicial
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),  # ayuda a reducir el sobreajuste, ver clase 4
    tf.keras.layers.Dense(10, activation='softmax')
])

# 4. Compilación: clasificación multiclase con etiquetas enteras (no one-hot)
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Entrenamiento con conjunto de validación reservado automáticamente
history = model.fit(
    train_images, train_labels,
    epochs=10,
    validation_split=0.2,
    verbose=0
)

# 6. Curvas de entrenamiento frente a validación
plt.plot(history.history['accuracy'], label='Precisión de entrenamiento')
plt.plot(history.history['val_accuracy'], label='Precisión de validación')
plt.xlabel('Época')
plt.ylabel('Precisión')
plt.legend()
plt.title('Curvas de entrenamiento vs. validación')
plt.show()

# Si la precisión de entrenamiento sigue subiendo mientras la de validación se estanca
# o empeora, el modelo está empezando a sobreajustar: ese es exactamente el patrón
# que aprendiste a detectar en la clase 4.

# 7. Guardado del modelo entrenado, listo para reutilizar o convertir a TensorFlow Lite
model.save('mi_proyecto_final.keras')
print('Modelo guardado correctamente')
```
