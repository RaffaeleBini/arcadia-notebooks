## Parte 1 — Fondamenti di Machine Learning

### Programmazione tradizionale rispetto a Machine Learning

Pensa al gioco di sassi, carta o forbici ("Rock, Paper, Scissors"). Riconoscere il gesto è facile per una persona, ma programmarlo con regole esplicite è molto difficile: ogni mano, ogni tono di pelle e ogni modo di fare il gesto sono diversi.

- **Programmazione tradizionale**: regole (codice) + dati → risultato. Tu scrivi le regole manualmente
- **Machine Learning**: dati + risultato → regole. Dai al sistema esempi già etichettati e lascia che deduca il modello

### Un esempio più semplice: numeri con una relazione nascosta

Prima di riconoscere gesti, pratichiamo con qualcosa di semplice: alcune coppie di numeri X e Y con una relazione nascosta, ad esempio Y = 2X - 1. Nella programmazione tradizionale scriveresti direttamente quella formula. In ML, lasciamo che il modello la deduca dai dati stessi.

### Che cos'è un modello e cosa significa addestrarlo

Un **modello** è la rappresentazione matematica che collega ingressi e uscite. All'inizio, i suoi parametri sono quasi casuali: la sua prima "ipotesi" potrebbe essere Y = 5X + 5.

**Addestrare** significa ripetere un ciclo:

1. Il modello predice
2. Una **funzione di perdita (loss)** misura quanto si sbaglia
3. Un **ottimizzatore** regola i parametri per ridurre quell'errore
4. Il ciclo si ripete per varie **epoche (epochs)**

I **dati di addestramento** sono gli esempi che il modello usa per imparare. Con pochi esempi, le previsioni del modello saranno approssimazioni probabili, non certezze esatte — ed è normale nel Machine Learning.

## Parte 2 — Primi passi con TensorFlow e Keras

### Che cos'è un tensor

Un **tensor** è un array multidimensionale con un tipo di dato (`dtype`) e una forma (`shape`). TensorFlow li utilizza come struttura centrale perché ha bisogno di un formato unico—da un singolo numero a un lotto di migliaia di immagini—su cui applicare le stesse operazioni ottimizzate su CPU (Central Processing Unit) o GPU (Graphics Processing Unit).

Verifichiamolo con il codice.

```python
import tensorflow as tf

print("Versión de TensorFlow:", tf.__version__)
```

```python
# Operaciones matemáticas básicas sobre tensores
print(tf.math.add(1, 2))                  # suma
print(tf.math.add([1, 2], [3, 4]))        # suma de vectores, elemento a elemento
print(tf.math.square(5))                  # elevar al cuadrado
print(tf.math.reduce_sum([1, 2, 3]))      # suma de todos los elementos

# TensorFlow sobrecarga los operadores habituales de Python
print(tf.math.square(2) + tf.math.square(3))
```

```python
# Cada tensor tiene una forma (shape) y un tipo de dato (dtype)
x = tf.linalg.matmul([[1]], [[2, 3]])  # multiplicación de matrices
print(x)
print("shape:", x.shape)
print("dtype:", x.dtype)
```

```python
import numpy as np

ndarray = np.ones([3, 3])

# TensorFlow convierte automáticamente un ndarray de NumPy en tensor
tensor = tf.math.multiply(ndarray, 42)
print("Tensor:\n", tensor)

# NumPy también convierte automáticamente un tensor en ndarray
print("Con NumPy:\n", np.add(tensor, 1))

# El método .numpy() convierte un tensor en un array de NumPy de forma explícita
print("Con .numpy():\n", tensor.numpy())
```

### La tua prima rete neurale: un solo neurone

Costruiamo l'esempio dei numeri in codice reale. Vogliamo che il modello apprenda la relazione Y = 2X - 1 partendo da appena sei coppie di dati, senza dirgli la formula.

**Passo 1 — Definire il modello.** Uno strato `Dense` (denso, completamente connesso) con un solo neurone (`units=1`) che riceve un unico valore in ingresso.

```python
from tensorflow import keras

model = tf.keras.Sequential([keras.layers.Dense(units=1, input_shape=[1])])
```

**Passo 2 — Compilare il modello.** Indichiamo la funzione di perdita (`loss`) e l'ottimizzatore:

- `mean_squared_error`: misura l'errore quadratico medio tra la previsione e il valore reale
- `sgd` (Stochastic Gradient Descent, discesa del gradiente stocastica): aggiusta i parametri per ridurre quell'errore

```python
model.compile(optimizer='sgd', loss='mean_squared_error')
```

**Passo 3 — Fornire i dati di addestramento.** Le stesse coppie X-Y di prima, come array di NumPy.

```python
xs = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-2.0, 1.0, 4.0, 7.0, 10.0, 13.0], dtype=float)
```

**Passo 4 — Addestrare con `fit`.** Osserva come scende la perdita (`loss`) epoca dopo epoca: questo è il modello che apprende.

```python
model.fit(xs, ys, epochs=500)
```

### La tua prima rete neurale: un singolo neurone

Andiamo a costruire l'esempio dei numeri in codice reale. Vogliamo che il modello impari la relazione Y = 2X - 1 da sole sei coppie di dati, senza dirgli la formula.

**Passo 1 — Definire il modello.** Uno strato `Dense` (denso, completamente connesso) con un singolo neurone (`units=1`) che riceve un singolo valore di input.

```python
print(model.predict(np.array([10.0])))
```

Il risultato sarà molto vicino a 19, ma probabilmente non esatto (qualcosa come 18,9998). Non è un errore: il modello lavora con probabilità apprese da appena sei esempi, non con certezze assolute.

### Uno sguardo a un modello più complesso

Lo stesso schema —strati, `compile`, `fit`, `predict`— si applica a modelli molto più grandi. Questo è un classificatore di cifre scritte a mano (dataset MNIST) che vedrai in dettaglio nella prossima lezione: qui limitati a notare che la struttura del codice è la stessa.

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

modelo_mnist = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),   # aplana la imagen 28x28 en un vector
    tf.keras.layers.Dense(128, activation='relu'),   # capa densa de 128 neuronas
    tf.keras.layers.Dropout(0.2),                    # apaga el 20% de las neuronas al azar
    tf.keras.layers.Dense(10)                        # una salida por cada dígito (0-9)
])

modelo_mnist.compile(optimizer='adam',
                      loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
                      metrics=['accuracy'])

modelo_mnist.fit(x_train, y_train, epochs=5)
modelo_mnist.evaluate(x_test, y_test, verbose=2)
```

**Passo 2 — Compilare il modello.** Indichiamo la funzione di perdita (`loss`) e l'ottimizzatore:

- `mean_squared_error`: misura l'errore quadratico medio tra la predizione e il valore reale
- `sgd` (Stochastic Gradient Descent, discesa del gradiente stocastica): regola i parametri per ridurre quell'errore
