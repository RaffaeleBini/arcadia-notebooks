def clasificar(frases, modelo):
    probabilidades = modelo.predict(frases).flatten()
    etiquetas_predichas = []
    for prob in probabilidades:
        etiquetas_predichas.append("positiva" if prob >= 0.5 else "negativa")
    return etiquetas_predichas
