## Esercizio 2 — Aggiungi dropout per ridurre l'overfitting

**Enunciato**: aggiungi uno strato `Dropout(0.5)` appena prima dello strato di output, affinché la rete "spenga" a caso metà dei neuroni durante l'addestramento e generalizzi meglio.

**Suggerimento**: `Dropout` influisce solo sull'addestramento; nella valutazione si usano tutti i neuroni.


````python
modelo_con_dropout = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Flatten(),
    # TODO: añade una capa Dropout(0.5) aquí
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_con_dropout.compile(optimizer="adam",
                           loss="sparse_categorical_crossentropy",
                           metrics=["accuracy"])

modelo_con_dropout.fit(x_entreno, y_entreno, epochs=3)
modelo_con_dropout.evaluate(x_test, y_test)
````
