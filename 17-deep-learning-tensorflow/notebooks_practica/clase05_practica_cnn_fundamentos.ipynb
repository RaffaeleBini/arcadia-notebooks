## Ejercicio 2 — Añade dropout para reducir el sobreajuste

**Enunciado**: añade una capa `Dropout(0.5)` justo antes de la capa de salida, para que la red "apague" al azar la mitad de las neuronas durante el entrenamiento y generalice mejor.

**Pista**: `Dropout` solo afecta al entrenamiento; en la evaluación se usan todas las neuronas.

````python
modelo_con_dropout = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Flatten(),
    # TODO: añade una capa Dropout(0.5) aquí
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_con_dropout.compile(optimizer="adam",
                           loss="sparse_categorical_crossentropy",
                           metrics=["accuracy"])

modelo_con_dropout.fit(x_entreno, y_entreno, epochs=3)
modelo_con_dropout.evaluate(x_test, y_test)
````
