## Esercizio 2 — Adatta il modello capstone

**Enunciato**: nel notebook di classe hai combinato bag of words (capitolo 2) con uno strato `LSTM` (capitolo 3). Sostituisci lo strato `LSTM` con uno strato `GRU` delle stesse unità e addestra il modello con gli stessi dati.

**Suggerimento**: `GRU` si importa allo stesso modo di `LSTM`, da `tf.keras.layers`.


````python
import tensorflow as tf
import numpy as np

with open("../pos.txt", encoding="latin-1") as f:
    positivas = [linea.strip() for linea in f if linea.strip()]
with open("../neg.txt", encoding="latin-1") as f:
    negativas = [linea.strip() for linea in f if linea.strip()]

textos = np.array(positivas + negativas)
etiquetas = np.array([1] * len(positivas) + [0] * len(negativas))

vectorizador = tf.keras.layers.TextVectorization(max_tokens=5000, output_sequence_length=30)
vectorizador.adapt(textos)

modelo_gru = tf.keras.Sequential([
    vectorizador,
    tf.keras.layers.Embedding(input_dim=5000, output_dim=32),
    # TODO: sustituye esta línea por una capa GRU(32)
    tf.keras.layers.LSTM(32),
    tf.keras.layers.Dense(1, activation="sigmoid"),
])

modelo_gru.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
modelo_gru.fit(textos, etiquetas, epochs=3)
````
