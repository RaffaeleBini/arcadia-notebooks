## Esercizio 2 — Converti il tuo proprio modello

**Enunciato**: addestra il modello `modelo_practica` della cella successiva, salvalo come `"modelo_practica.h5"`, e completa il comando di conversione per generare la cartella `modelo_web_practica/`.

**Suggerimento**: il formato di input di Keras si indica con `--input_format=keras`.


````python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_entreno = x_entreno / 255.0

modelo_practica = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax"),
])
modelo_practica.compile(optimizer="adam", loss="sparse_categorical_crossentropy")
modelo_practica.fit(x_entreno, y_entreno, epochs=1)
# TODO: guarda el modelo como "modelo_practica.h5"
````
