## 2. Mini capstone: combina due tecniche in un'unica pipeline

Combinerai due blocchi del corso: il classificatore di sentimenti (capitolo 2) e uno strato ricorrente (capitolo 3), al posto del bag of words originale. Es un ejemplo de cómo un proyecto real casi nunca usa una sola técnica aislada.

```python
import tensorflow as tf
import numpy as np

with open("../pos.txt", encoding="latin-1") as f:
    positivas = [linea.strip() for linea in f if linea.strip()]
with open("../neg.txt", encoding="latin-1") as f:
    negativas = [linea.strip() for linea in f if linea.strip()]

textos = np.array(positivas + negativas)
etiquetas = np.array([1] * len(positivas) + [0] * len(negativas))
indices = np.arange(len(textos))
np.random.seed(0)
np.random.shuffle(indices)
textos, etiquetas = textos[indices], etiquetas[indices]
corte = int(len(textos) * 0.8)
textos_entreno, textos_test = textos[:corte], textos[corte:]
etiquetas_entreno, etiquetas_test = etiquetas[:corte], etiquetas[corte:]
```

```python
LONGITUD_MAXIMA = 30

vectorizador = tf.keras.layers.TextVectorization(
    max_tokens=5000,
    output_sequence_length=LONGITUD_MAXIMA,
)
vectorizador.adapt(textos_entreno)

# Capítulo 2 (bolsa de palabras) + capítulo 3 (memoria secuencial): combinamos ambos enfoques
modelo_capstone = tf.keras.Sequential([
    vectorizador,
    tf.keras.layers.Embedding(input_dim=5000, output_dim=32),
    tf.keras.layers.LSTM(32),
    tf.keras.layers.Dense(1, activation="sigmoid"),
])

modelo_capstone.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
modelo_capstone.fit(textos_entreno, etiquetas_entreno, epochs=5, validation_split=0.1)
modelo_capstone.evaluate(textos_test, etiquetas_test)
```

A differenza del capitolo 2, qui lo strato `Embedding` + `LSTM` tiene conto dell'**ordine** delle parole, non solo di quali appaiono. Confronta l'accuratezza finale con quella del capitolo 2: migliora con questo dataset?
