## 1. Prepara i dati (li conosci già)

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0
```

## 2. Definisci il modello con Keras (alto livello)

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])
```

## 3. Aggiungi callback: interrompi l'addestramento e salva il miglior modello automaticamente

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True),
    tf.keras.callbacks.ModelCheckpoint("mejor_modelo.keras", save_best_only=True),
]

historial = modelo.fit(x_entreno, y_entreno,
                       epochs=20,
                       validation_split=0.1,
                       callbacks=callbacks)

print("Número de épocas realmente entrenadas:", len(historial.history["loss"]))
```

## 4. Valuta il modello finale

```python
modelo.evaluate(x_test, y_test)
```

## Riepilogo

- Hai definito una rete con Keras in poche righe, senza gestire tu stesso i pesi o i gradienti.
- `EarlyStopping` ha interrotto l'addestramento non appena la metrica di validazione ha smesso di migliorare, evitando l'overfitting.
- `ModelCheckpoint` ha salvato automaticamente la migliore versione del modello vista durante l'addestramento.
