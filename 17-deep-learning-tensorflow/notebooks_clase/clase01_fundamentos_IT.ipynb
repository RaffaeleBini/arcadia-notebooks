## 1. Cos'è una rete neurale?

Immagina una commissione di dieci persone che vota se una foto mostra un gatto: ciascuno si concentra su un indizio diverso (orecchi, baffi, occhi) e dà un voto ponderato. Una rete neurale fa lo stesso con "neuroni" organizzati in strati: ogni strato combina i voti ponderati (pesi) dello strato precedente fino a raggiungere una risposta finale.

L'apprendimento profondo (Deep Learning, DL) impila molti strati per riconoscere pattern complessi senza programmare regole manualmente. TensorFlow è lo strumento che ti consente di costruire, addestrare e usare quelle reti.

## 2. Verifica la tua installazione di TensorFlow

Esegui la cella seguente. Se vedi il messaggio stampato senza errori, tutto è pronto.

```python
import tensorflow as tf

print("Versión de TensorFlow:", tf.__version__)
print(tf.constant("¡Hola, TensorFlow!"))
```

## 3. Carica e prepara il dataset MNIST

MNIST contiene settantamila immagini di cifre scritte a mano (0-9), in bianco e nero, di 28x28 pixel.

```python
(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalizamos los píxeles a un rango de 0 a 1 para que la red aprenda mejor
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0

print("Imágenes de entrenamiento:", x_entreno.shape)
print("Imágenes de test:", x_test.shape)
```

## 4. Visualizza una cifra di esempio

Utile per mostrare in classe l'aspetto dei dati prima dell'addestramento.

```python
import matplotlib.pyplot as plt

plt.imshow(x_entreno[0], cmap="gray")
plt.title(f"Dígito real: {y_entreno[0]}")
plt.axis("off")
plt.show()
```

## 5. Definisci e addestra la tua prima rete con Keras

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])

modelo.summary()
```

```python
historial = modelo.fit(x_entreno, y_entreno, epochs=5, validation_split=0.1)
```

## 6. Valuta il modello con dati che non ha mai visto

```python
perdida_test, exactitud_test = modelo.evaluate(x_test, y_test)
print(f"Exactitud sobre el conjunto de test: {exactitud_test:.2%}")
```
