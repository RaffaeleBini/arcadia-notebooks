## 1. De red densa a red convolucional

Una capa convolucional desliza un filtro pequeño (por ejemplo 3x3) por toda la imagen, detectando patrones locales como bordes o texturas. Una capa de pooling resume esa información quedándose con el valor más relevante de cada bloque.

## 2. Prepara los datos con el canal de color explícito

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_entreno = (x_entreno / 255.0).reshape(-1, 28, 28, 1)
x_test = (x_test / 255.0).reshape(-1, 28, 28, 1)

print("Forma de entrada:", x_entreno.shape)
```

## 3. Define la CNN: convolución, pooling, convolución, pooling

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Conv2D(64, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])

modelo.summary()
```

```python
modelo.fit(x_entreno, y_entreno, epochs=5, validation_split=0.1)
modelo.evaluate(x_test, y_test)
```

## 4. Visualiza un filtro de la primera capa convolucional

Útil para mostrar en clase qué aprende la red: cada filtro reacciona a un patrón distinto.

```python
import matplotlib.pyplot as plt

primer_filtro = modelo.layers[0].get_weights()[0][:, :, 0, 0]
plt.imshow(primer_filtro, cmap="gray")
plt.title("Un filtro de la primera capa convolucional")
plt.colorbar()
plt.show()
```

## Resumen

- Construiste una CNN con dos bloques de convolución + pooling.
- La red suele superar el 99% de exactitud en MNIST, mejor que las redes densas de los capítulos anteriores.
- El patrón convolución → pooling → convolución → pooling → aplanado → densa se repite en la inmensa mayoría de las CNN.
