## 1. ¿Qué es una red neuronal?

Piensa en un comité de diez personas votando si una foto muestra un gato: cada una se fija en una pista distinta (orejas, bigotes, ojos) y da un voto ponderado. Una red neuronal hace lo mismo con "neuronas" organizadas en capas: cada capa combina los votos ponderados (pesos) de la anterior hasta llegar a una respuesta final.

El aprendizaje profundo (Deep Learning, DL) apila muchas capas para reconocer patrones complejos sin programar reglas a mano. TensorFlow es la herramienta que te permite construir, entrenar y usar esas redes.

## 3. Carga y prepara el dataset MNIST

MNIST contiene setenta mil imágenes de dígitos escritos a mano (0-9), en blanco y negro, de 28x28 píxeles.

```python
(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalizamos los píxeles a un rango de 0 a 1 para que la red aprenda mejor
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0

print("Imágenes de entrenamiento:", x_entreno.shape)
print("Imágenes de test:", x_test.shape)
```

## 4. Proyecta un dígito de ejemplo

Útil para mostrar en clase qué aspecto tienen los datos antes de entrenar.

```python
import matplotlib.pyplot as plt

plt.imshow(x_entreno[0], cmap="gray")
plt.title(f"Dígito real: {y_entreno[0]}")
plt.axis("off")
plt.show()
```

## 5. Define y entrena tu primera red con Keras

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])

modelo.summary()
```

```python
historial = modelo.fit(x_entreno, y_entreno, epochs=5, validation_split=0.1)
```

## 6. Evalúa el modelo con datos que nunca ha visto

```python
perdida_test, exactitud_test = modelo.evaluate(x_test, y_test)
print(f"Exactitud sobre el conjunto de test: {exactitud_test:.2%}")
```
