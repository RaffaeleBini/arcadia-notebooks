## 1. Da rete densa a rete convoluzionale

Uno strato convoluzionale fa scorrere un filtro piccolo (ad esempio 3x3) su tutta l'immagine, rilevando pattern locali come bordi o trame. Uno strato di pooling riassume quell'informazione mantenendo il valore più rilevante di ogni blocco.

## 2. Prepara i dati con il canale di colore esplicito

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_entreno = (x_entreno / 255.0).reshape(-1, 28, 28, 1)
x_test = (x_test / 255.0).reshape(-1, 28, 28, 1)

print("Forma de entrada:", x_entreno.shape)
```

## 3. Definisci la CNN: convoluzione, pooling, convoluzione, pooling

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Conv2D(64, kernel_size=3, activation="relu"),
    tf.keras.layers.MaxPooling2D(pool_size=2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])

modelo.summary()
```

```python
modelo.fit(x_entreno, y_entreno, epochs=5, validation_split=0.1)
modelo.evaluate(x_test, y_test)
```

## 4. Visualizza un filtro del primo strato convoluzionale

Utile per mostrare in classe cosa apprende la rete: ogni filtro reagisce a un pattern diverso.

```python
import matplotlib.pyplot as plt

primer_filtro = modelo.layers[0].get_weights()[0][:, :, 0, 0]
plt.imshow(primer_filtro, cmap="gray")
plt.title("Un filtro de la primera capa convolucional")
plt.colorbar()
plt.show()
```

## Riepilogo

- Hai costruito una CNN con due blocchi di convoluzione + pooling.
- La rete di solito supera il 99% di precisione su MNIST, meglio delle reti dense dei capitoli precedenti.
- Il pattern convoluzione → pooling → convoluzione → pooling → appiattito → denso si ripete nella stragrande maggioranza delle CNN.
