**Objetivos de hoy**
- Reconocer las ventajas de una API de alto nivel frente a escribir TensorFlow "a pelo".
- Usar callbacks de Keras para automatizar buenas prácticas de entrenamiento.

Consulta el [capítulo 5 del manual](../manual/05_tflearn_keras.md) para la explicación completa.

## 1. Prepara los datos (ya los conoces)

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0
```

## 2. Define el modelo con Keras (alto nivel)

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo.compile(optimizer="adam",
               loss="sparse_categorical_crossentropy",
               metrics=["accuracy"])
```

## 3. Añade callbacks: detén el entrenamiento y guarda el mejor modelo automáticamente

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True),
    tf.keras.callbacks.ModelCheckpoint("mejor_modelo.keras", save_best_only=True),
]

historial = modelo.fit(x_entreno, y_entreno,
                       epochs=20,
                       validation_split=0.1,
                       callbacks=callbacks)

print("Número de épocas realmente entrenadas:", len(historial.history["loss"]))
```

## 4. Evalúa el modelo final

```python
modelo.evaluate(x_test, y_test)
```

## Resumen

- Definiste una red con Keras en unas pocas líneas, sin gestionar tú mismo pesos ni gradientes.
- `EarlyStopping` detuvo el entrenamiento en cuanto la métrica de validación dejó de mejorar, evitando entrenar de más.
- `ModelCheckpoint` guardó automáticamente la mejor versión del modelo visto durante el entrenamiento.
