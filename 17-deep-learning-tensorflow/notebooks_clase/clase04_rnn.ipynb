**Objetivos de hoy**
- Entender por qué una red necesita "memoria" para procesar secuencias.
- Distinguir una celda LSTM de una celda recurrente simple.
- Aplicar una LSTM a MNIST tratado como secuencia de filas.

Consulta el [capítulo 3 del manual](../manual/03_rnn_lstm.md) para la explicación completa.

## 1. ¿Por qué una red "con memoria"?

Cuando lees una frase, el significado de cada palabra depende de las anteriores. Una red densa trata cada entrada de forma aislada; una RNN mantiene un "estado oculto" que resume lo visto hasta el momento. LSTM y GRU son variantes que evitan que ese resumen "olvide" información lejana en la secuencia.

## 2. Prepara MNIST como una secuencia

Cada imagen de 28x28 se interpreta como una secuencia de 28 pasos de tiempo, cada uno con un vector de 28 valores (una fila de píxeles).

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0

print("Forma de cada imagen-secuencia:", x_entreno.shape[1:])
```

## 3. Define y entrena una red con una capa LSTM

```python
modelo_lstm = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_lstm.compile(optimizer="adam",
                    loss="sparse_categorical_crossentropy",
                    metrics=["accuracy"])

modelo_lstm.summary()
```

```python
modelo_lstm.fit(x_entreno, y_entreno, epochs=3, validation_split=0.1)
modelo_lstm.evaluate(x_test, y_test)
```

## 4. Compara con una capa GRU

GRU tiene menos puertas internas que LSTM, y por tanto menos parámetros que entrenar.

```python
modelo_gru = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.GRU(128),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_gru.compile(optimizer="adam",
                   loss="sparse_categorical_crossentropy",
                   metrics=["accuracy"])

modelo_gru.fit(x_entreno, y_entreno, epochs=3, validation_split=0.1)
modelo_gru.evaluate(x_test, y_test)
```

## Resumen

- Una LSTM procesa cada imagen como una secuencia de filas, manteniendo un estado que resume lo visto hasta el momento.
- LSTM y GRU suelen rendir de forma parecida; GRU es algo más ligera.
- Cambiar de una arquitectura densa a una recurrente en Keras es tan sencillo como sustituir una capa.
