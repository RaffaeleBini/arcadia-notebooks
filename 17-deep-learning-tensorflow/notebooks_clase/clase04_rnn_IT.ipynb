## 1. Perché una rete "con memoria"?

Quando leggi una frase, il significato di ogni parola dipende da quelle precedenti. Una rete densa tratta ogni input isolatamente; una RNN mantiene uno "stato nascosto" che riassume ciò che è stato visto fino al momento. LSTM e GRU sono varianti che impediscono a quel riepilogo di "dimenticare" informazioni lontane nella sequenza.

## 2. Prepara MNIST come una sequenza

Ogni immagine di 28x28 è interpretata come una sequenza di 28 step temporali, ciascuno con un vettore di 28 valori (una riga di pixel).

```python
import tensorflow as tf

(x_entreno, y_entreno), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_entreno, x_test = x_entreno / 255.0, x_test / 255.0

print("Forma de cada imagen-secuencia:", x_entreno.shape[1:])
```

## 3. Definisci e addestra una rete con uno strato LSTM

```python
modelo_lstm = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_lstm.compile(optimizer="adam",
                    loss="sparse_categorical_crossentropy",
                    metrics=["accuracy"])

modelo_lstm.summary()
```

```python
modelo_lstm.fit(x_entreno, y_entreno, epochs=3, validation_split=0.1)
modelo_lstm.evaluate(x_test, y_test)
```

## 4. Confronta con uno strato GRU

GRU ha meno gate interni rispetto a LSTM, e quindi meno parametri da addestrare.

```python
modelo_gru = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.GRU(128),
    tf.keras.layers.Dense(10, activation="softmax"),
])

modelo_gru.compile(optimizer="adam",
                   loss="sparse_categorical_crossentropy",
                   metrics=["accuracy"])

modelo_gru.fit(x_entreno, y_entreno, epochs=3, validation_split=0.1)
modelo_gru.evaluate(x_test, y_test)
```

## Riepilogo

- Una LSTM elabora ogni immagine come una sequenza di righe, mantenendo uno stato che riassume ciò che è stato visto fino al momento.
- LSTM e GRU di solito funzionano in modo simile; GRU è un po' più leggera.
- Passare da un'architettura densa a una ricorrente in Keras è semplice come sostituire uno strato.
