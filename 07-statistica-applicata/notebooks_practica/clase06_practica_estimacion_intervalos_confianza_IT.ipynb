### Soluzione dell'esercizio 1

```python
tiempos = np.array([42, 38, 51, 45, 40, 47, 39, 44, 50])

n = len(tiempos)
media = tiempos.mean()
s = tiempos.std(ddof=1)
ee = s / np.sqrt(n)
t_crit = stats.t.ppf(0.975, df=n - 1)
ic_inf = media - t_crit * ee
ic_sup = media + t_crit * ee

assert all(v is not None for v in (n, media, s, ee, t_crit, ic_inf, ic_sup)), "Completa le righe contrassegnate con TODO"
assert n == 9 and np.isclose(media, 44.0), "rivedi n e la media"
assert np.isclose(s, np.sqrt(22)), "s usa ddof=1: la somma dei quadrati è 176 e 176/8 = 22"
assert np.isclose(ee, s / 3), "l'errore standard è s / √n"
assert np.isclose(t_crit, 2.306, atol=0.001), "usa il quantile 0,975 con n - 1 = 8 gradi di libertà"
ref = stats.t.interval(0.95, df=8, loc=44.0, scale=np.sqrt(22) / 3)
assert np.isclose(ic_inf, ref[0]) and np.isclose(ic_sup, ref[1]), "l'IC è media ± t_crit · ee"
print(f"IC 95 % = ({ic_inf:.2f}, {ic_sup:.2f})")
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 2

```python
def semiancho(s, n, nivel):
    """Semi-ampiezza dell'IC per la media: t · s / √n."""
    t_crit = stats.t.ppf(1 - (1 - nivel) / 2, df=n - 1)
    return t_crit * s / np.sqrt(n)

sa_90 = semiancho(4.69, 9, 0.90)
sa_99 = semiancho(4.69, 9, 0.99)
factor_n4 = semiancho(4.69, 36, 0.95) / semiancho(4.69, 9, 0.95)

assert semiancho(4.69, 9, 0.95) is not None, "Completa la funzione semiancho (TODO)"
assert all(v is not None for v in (sa_90, sa_99, factor_n4)), "Completa le righe contrassegnate con TODO"
assert np.isclose(semiancho(4.69, 9, 0.95), 2.306 * 4.69 / 3, rtol=1e-3), "t(0,975; 8) = 2,306"
assert sa_90 < semiancho(4.69, 9, 0.95) < sa_99, "più confidenza, più ampio"
assert np.isclose(sa_99 / sa_90, 3.355 / 1.860, rtol=1e-2), "rivedi i quantili t del 90 % e del 99 %"
assert 0.43 < factor_n4 < 0.45, "con n quattro volte maggiore il rapporto è un po' inferiore a 0,5 (per la t)"
print(f"sa_90 = {sa_90:.3f}   sa_99 = {sa_99:.3f}   factor_n4 = {factor_n4:.3f}")
print("Esercizio 2 corretto ✔")
```

### Soluzione dell'esercizio 3

```python
def cobertura(n, nivel, reps, rng, usar_t):
    """Proporzione di IC che contengono la media vera (50)."""
    m = rng.normal(50, 10, size=(reps, n))
    ees = m.std(axis=1, ddof=1) / np.sqrt(n)
    q = 1 - (1 - nivel) / 2
    crit = stats.t.ppf(q, df=n - 1) if usar_t else stats.norm.ppf(q)
    medias = m.mean(axis=1)
    return ((medias - crit * ees <= 50) & (50 <= medias + crit * ees)).mean()

rng_ej = np.random.default_rng(123)
cob_t = cobertura(5, 0.95, 1_000, rng_ej, True)
cob_z = cobertura(5, 0.95, 1_000, rng_ej, False)

assert cob_t is not None, "Completa la funzione cobertura (TODO)"
assert 0.92 < cob_t < 0.98, "la t con n = 5 dovrebbe coprire vicino a 0,95"
assert cob_z < 0.92, "con z e n = 5 la copertura dovrebbe essere nettamente minore (≈ 0,88)"
assert cob_z < cob_t - 0.03, "z deve coprire meno di t"
print(f"copertura con t = {cob_t:.3f}   copertura con z = {cob_z:.3f}")
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
datos = np.random.default_rng(11).lognormal(mean=np.log(30), sigma=0.5, size=30)     # 30 tempi di consegna (esempio illustrativo)

def bootstrap_ic(datos, estadistico, nivel, B, rng):
    """IC percentile bootstrap: restituisce (inferior, superior)."""
    idx = rng.integers(0, len(datos), size=(B, len(datos)))
    dist = estadistico(datos[idx], axis=1)
    alfa = 1 - nivel
    inf, sup = np.quantile(dist, [alfa / 2, 1 - alfa / 2])
    return inf, sup

rng_ej = np.random.default_rng(5)
ic_mediana = bootstrap_ic(datos, np.median, 0.95, 5_000, rng_ej)

assert ic_mediana is not None, "Completa la funzione bootstrap_ic (TODO)"
inferior, superior = ic_mediana
assert inferior < np.median(datos) < superior, "l'IC dovrebbe contenere la mediana del campione"
ref = np.quantile(np.median(datos[np.random.default_rng(999).integers(0, 30, size=(20_000, 30))], axis=1), [0.025, 0.975])
assert abs(inferior - ref[0]) < 1.5 and abs(superior - ref[1]) < 1.5, "i tuoi limiti si discostano dal riferimento: ricampioni con reinserimento e di dimensione n?"
print(f"Mediana campionaria = {np.median(datos):.2f}   IC bootstrap 95 % = ({inferior:.2f}, {superior:.2f})")
print("Esercizio 4 corretto ✔")
```

### Soluzione della sfida finale

```python
datos_reto = np.random.default_rng(2026).lognormal(mean=np.log(30), sigma=0.5, size=200)

def informe_ic(datos, nivel, B, rng):
    """IC della media (t), della media (bootstrap) e della mediana (bootstrap)."""
    idx = rng.integers(0, len(datos), size=(B, len(datos)))
    remuestras = datos[idx]
    alfa = 1 - nivel
    cuantiles = [alfa / 2, 1 - alfa / 2]
    return {"n": len(datos),
            "t": stats.t.interval(nivel, df=len(datos) - 1, loc=datos.mean(), scale=stats.sem(datos)),
            "boot_media": tuple(np.quantile(remuestras.mean(axis=1), cuantiles)),
            "boot_mediana": tuple(np.quantile(np.median(remuestras, axis=1), cuantiles))}

res = informe_ic(datos_reto, 0.95, 10_000, np.random.default_rng(1))

assert res is not None, "Completa la funzione informe_ic (TODO)"
assert res["n"] == 200, "n è la dimensione del campione"
ref_t = stats.t.interval(0.95, df=199, loc=datos_reto.mean(), scale=stats.sem(datos_reto))
assert np.allclose(res["t"], ref_t), "rivedi l'IC con la t (n - 1 gradi di libertà, errore standard s/√n)"
ancho_t = res["t"][1] - res["t"][0]
ancho_b = res["boot_media"][1] - res["boot_media"][0]
assert abs(ancho_b / ancho_t - 1) < 0.15, "con n = 200 l'IC bootstrap della media deve somigliare a quello della t"
assert res["boot_media"][0] < datos_reto.mean() < res["boot_media"][1], "l'IC bootstrap deve contenere la media campionaria"
assert res["boot_mediana"][0] < np.median(datos_reto) < res["boot_mediana"][1], "l'IC della mediana deve contenere la mediana campionaria"
print(pd.DataFrame({k: v for k, v in res.items() if k != "n"}, index=["inferiore", "superiore"]).round(2))
print("Sfida finale corretta ✔")
```
