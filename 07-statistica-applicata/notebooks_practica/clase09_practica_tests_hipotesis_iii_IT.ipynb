### Soluzione dell'esercizio 1

```python
tabla = np.array([[18, 82],
                  [32, 68]])

esperada = tabla.sum(axis=1, keepdims=True) * tabla.sum(axis=0, keepdims=True) / tabla.sum()
chi2_est = ((tabla - esperada) ** 2 / esperada).sum()
gl = (tabla.shape[0] - 1) * (tabla.shape[1] - 1)
p_chi = stats.chi2.sf(chi2_est, gl)
v_cramer = np.sqrt(chi2_est / (tabla.sum() * (min(tabla.shape) - 1)))
condicion_ok = bool(esperada.min() >= 5)

assert all(v is not None for v in (esperada, chi2_est, gl, p_chi, v_cramer, condicion_ok)), "Completa le righe contrassegnate con TODO"
assert np.allclose(esperada, [[25, 75], [25, 75]]), "attesa = totali di riga × totali di colonna / totale generale"
assert np.isclose(chi2_est, 5.2267, atol=1e-3), "χ² = somma di (osservata − attesa)² / attesa"
assert gl == 1, "gdl = (righe − 1) · (colonne − 1)"
assert np.isclose(p_chi, 0.0222, atol=1e-3), "p = stats.chi2.sf(chi2_est, gl)"
assert np.isclose(v_cramer, 0.1617, atol=1e-3), "V = radice di χ² / (N · (min(righe, colonne) − 1))"
assert condicion_ok is True or condicion_ok == True, "confronta la minima attesa con 5"
print(f"χ² = {chi2_est:.3f}   gl = {gl}   p = {p_chi:.4f}   V = {v_cramer:.3f}")
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
def bh_a_mano(pvalores, alfa=0.05):
    p = np.asarray(pvalores, dtype=float)
    m = len(p)
    orden = np.argsort(p)
    aj_orden = np.minimum.accumulate((p[orden] * m / np.arange(1, m + 1))[::-1])[::-1]
    aj_orden = np.minimum(aj_orden, 1.0)
    p_ajustado = np.empty(m)
    p_ajustado[orden] = aj_orden
    return p_ajustado <= alfa, p_ajustado

pv12 = np.array([0.0433, 0.0017, 0.4278, 0.016, 0.5457, 0.0047, 0.2102, 0.0373, 0.7505, 0.0238, 0.1669, 0.5103])
res_bh = bh_a_mano(pv12)
n_bh = int(res_bh[0].sum())

assert res_bh is not None and n_bh is not None, "Completa le righe contrassegnate con TODO"
rechaza, p_aj = res_bh
r_ref = multipletests(pv12, alpha=0.05, method="fdr_bh")
assert np.allclose(p_aj, r_ref[1]), "il p corretto deve coincidere con multipletests(..., method='fdr_bh')[1]"
assert np.array_equal(rechaza, r_ref[0]), "rechaza = p_ajustado <= alfa"
otro = np.random.default_rng(1).uniform(size=30) ** 2
assert np.allclose(bh_a_mano(otro)[1], multipletests(otro, method="fdr_bh")[1]), "la tua funzione deve funzionare con qualsiasi vettore di p-value"
assert n_bh == 2, "n_bh è il numero di True in rechaza"
print(f"BH rifiuta {n_bh} su {len(pv12)};  Bonferroni: {(pv12 <= 0.05 / len(pv12)).sum()};  senza correzione: {(pv12 < 0.05).sum()}")
print("Esercizio 4 corretto ✔")
```

```python
rng_r = np.random.default_rng(910)
tablas = []
for k, n in enumerate([150, 150, 150, 150, 150, 150, 20, 20, 15, 15]):
    a = rng_r.binomial(n, 0.20)
    b = rng_r.binomial(n, 0.34 if k in (0, 1) else 0.20)
    tablas.append(np.array([[a, n - a], [b, n - b]]))

def informe_tests(tablas):
    """Restituisce p, n_fisher, n_sin_corregir, n_bonf e n_bh."""
    p, n_fisher = [], 0
    for t in tablas:
        c = stats.chi2_contingency(t, correction=False)
        if c.expected_freq.min() < 5:
            p.append(stats.fisher_exact(t).pvalue)
            n_fisher += 1
        else:
            p.append(c.pvalue)
    p = np.array(p)
    return {"p": p, "n_fisher": n_fisher, "n_sin_corregir": int((p < 0.05).sum()),
            "n_bonf": int(multipletests(p, method="bonferroni")[0].sum()),
            "n_bh": int(multipletests(p, method="fdr_bh")[0].sum())}

inf = informe_tests(tablas)
respuesta = "b"

assert inf is not None and respuesta is not None, "Completa la funzione e la risposta (TODO)"
assert set(inf) == {"p", "n_fisher", "n_sin_corregir", "n_bonf", "n_bh"}, "il dizionario deve avere esattamente le chiavi p, n_fisher, n_sin_corregir, n_bonf e n_bh"
assert len(inf["p"]) == 10 and np.isclose(inf["p"][0], 0.0004, atol=1e-4) and np.isclose(inf["p"][7], 0.0471, atol=1e-3), "usa chi2_contingency(t, correction=False) se la minima attesa è ≥ 5 e fisher_exact(t) altrimenti"
assert inf["n_fisher"] == 3, "conta le tabelle con minima attesa < 5"
assert inf["n_sin_corregir"] == 3, "conta i p < 0,05"
assert inf["n_bonf"] == 2 and inf["n_bh"] == 2, "applica multipletests ai 10 p-value (bonferroni e fdr_bh)"
assert respuesta == "b", "il p-value non è la probabilità che H0 sia vera, e non correggere gonfia i falsi positivi"
print(f"tabelle con Fisher: {inf['n_fisher']}   rifiuti: senza correzione {inf['n_sin_corregir']}, Bonferroni {inf['n_bonf']}, BH {inf['n_bh']}")
print("Sfida finale corretta ✔")
```
