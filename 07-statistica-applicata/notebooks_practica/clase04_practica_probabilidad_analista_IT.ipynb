### Soluzione dell'esercizio 1

```python
from itertools import product

omega = list(product(range(1, 7), repeat=2))
p_suma_7 = sum(1 for a, b in omega if a + b == 7) / len(omega)
p_al_menos_un_6 = 1 - sum(1 for a, b in omega if a != 6 and b != 6) / len(omega)

assert len(omega) == 36, "lo spazio campionario ha 36 coppie"
assert np.isclose(p_suma_7, 6 / 36), "ci sono 6 coppie che sommano 7"
assert np.isclose(p_al_menos_un_6, 11 / 36), "P(almeno un 6) = 1 − 25/36"
print(f"P(somma = 7) = {p_suma_7:.4f}   P(almeno un 6) = {p_al_menos_un_6:.4f}")
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 2

### Soluzione dell'esercizio 3

```python
def vpp(prevalencia, sensibilidad, especificidad):
    """P(malato | positivo) con il teorema di Bayes."""
    p_positivo = prevalencia * sensibilidad + (1 - prevalencia) * (1 - especificidad)
    return prevalencia * sensibilidad / p_positivo

assert vpp(0.01, 0.95, 0.95) is not None, "Completa la funzione vpp (TODO)"
assert np.isclose(vpp(0.01, 0.95, 0.95), 95 / 590), "caso guida: 95 veri positivi su 590 positivi"
assert np.isclose(vpp(0.02, 0.90, 0.97), 0.018 / 0.0474), "rivedi la probabilità totale di risultare positivo"
assert np.isclose(vpp(0.50, 0.95, 0.95), 0.95), "con prevalenza 50 % e sensibilità = specificità, il VPP è uguale a loro"
assert vpp(0.001, 0.95, 0.95) < vpp(0.01, 0.95, 0.95) < vpp(0.1, 0.95, 0.95), "il VPP deve crescere con la prevalenza"
print(f"VPP(1 %, 95 %, 95 %) = {vpp(0.01, 0.95, 0.95):.4f}   VPP(2 %, 90 %, 97 %) = {vpp(0.02, 0.90, 0.97):.4f}")
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
def simular_vpp(prevalencia, sensibilidad, especificidad, n, rng):
    """Stima per simulazione di P(malato | positivo)."""
    enfermo = rng.random(n) < prevalencia
    u = rng.random(n)
    positivo = np.where(enfermo, u < sensibilidad, u < 1 - especificidad)
    return enfermo[positivo].mean()

rng_ej = np.random.default_rng(7)
estimacion = simular_vpp(0.05, 0.95, 0.95, 500_000, rng_ej)

assert estimacion is not None, "Completa la funzione simular_vpp (TODO)"
exacto = 0.05 * 0.95 / (0.05 * 0.95 + 0.95 * 0.05)          # = 0,5
n_pos_aprox = 500_000 * (0.05 * 0.95 + 0.95 * 0.05)
error_tipico = np.sqrt(exacto * (1 - exacto) / n_pos_aprox)
assert abs(estimacion - exacto) < 5 * error_tipico, "la stima è troppo lontana dal valore esatto: stai dividendo per i positivi?"
print(f"esatto = {exacto:.4f}   simulato = {estimacion:.4f}   (errore standard ≈ {error_tipico:.4f})")
print("Esercizio 4 corretto ✔")
```

### Soluzione dell'esercizio 5

```python
tiradas = rng.integers(1, 7, size=600_000)
previa_es_6 = tiradas[:-1] == 6      # lancio precedente
actual_es_6 = tiradas[1:] == 6       # lancio successivo

p_seis_tras_seis = actual_es_6[previa_es_6].mean()
p_dos_seises = (previa_es_6 & actual_es_6).mean()

assert None not in (p_seis_tras_seis, p_dos_seises), "Completa le righe contrassegnate con TODO"
assert abs(p_seis_tras_seis - 1 / 6) < 0.01, "se i dadi sono indipendenti, dovrebbe essere vicino a 1/6"
assert abs(p_dos_seises - 1 / 36) < 0.002, "dovrebbe essere vicino a 1/36"
print(f"P(6 | precedente 6) ≈ {p_seis_tras_seis:.4f} (1/6 = {1/6:.4f})   P(6 y 6) ≈ {p_dos_seises:.4f} (1/36 = {1/36:.4f})")
print("Esercizio 5 corretto ✔")
```

### Soluzione della sfida finale

```python
def informe(prevalencia, sensibilidad, especificidad, n=1_000_000):
    """Rapporto su un test: VPP e VPN esatti, VPP simulato e prevalenza minima per VPP ≥ 50 %."""
    p_pos = prevalencia * sensibilidad + (1 - prevalencia) * (1 - especificidad)
    p_neg = 1 - p_pos
    vpp_e = prevalencia * sensibilidad / p_pos
    vpn_e = (1 - prevalencia) * especificidad / p_neg
    enfermo = rng.random(n) < prevalencia
    u = rng.random(n)
    positivo = np.where(enfermo, u < sensibilidad, u < 1 - especificidad)
    return {"vpp": vpp_e, "vpn": vpn_e, "vpp_sim": enfermo[positivo].mean(),
            "prev_50": (1 - especificidad) / (sensibilidad + 1 - especificidad)}

res = informe(0.005, 0.90, 0.98)

assert res is not None, "Completa la funzione informe (TODO)"
assert np.isclose(res["vpp"], 0.0045 / 0.0244), "rivedi il VPP esatto"
assert np.isclose(res["vpn"], 0.9751 / 0.9756), "VPN = P(sano | negativo)"
assert np.isclose(res["prev_50"], 0.02 / 0.92), "prev = (1 − spec) / (sens + 1 − spec)"
assert abs(res["vpp_sim"] - res["vpp"]) < 0.015, "la simulazione dovrebbe avvicinarsi al VPP esatto"
print(pd.Series(res).round(4))
print("Sfida finale corretta ✔")
```
