### Solución del ejercicio 4

```python
def razon_dispersion(x):
    """Varianza muestral (ddof=1) dividida por la media."""
    return np.var(x, ddof=1) / np.mean(x)

def compatible_poisson(x, tol=0.25):
    """True si la razón varianza/media está a menos de tol de 1."""
    return bool(abs(razon_dispersion(x) - 1) < tol)

rng_ej = np.random.default_rng(5)
conteos_a = rng_ej.poisson(5, size=50_000)
conteos_b = rng_ej.negative_binomial(n=2.5, p=0.25, size=50_000)      # media 7,5 y varianza 30

assert razon_dispersion(np.array([1.0, 2.0, 3.0])) is not None, "Completa las funciones (TODO)"
assert np.isclose(razon_dispersion(np.array([1.0, 2.0, 3.0])), 0.5), "varianza(ddof=1) = 1 y media = 2 → razón 0,5"
assert 0.95 < razon_dispersion(conteos_a) < 1.05, "los datos Poisson deben dar una razón cercana a 1"
assert 3.5 < razon_dispersion(conteos_b) < 4.5, "los datos sobredispersos tienen varianza / media = 4"
assert compatible_poisson(conteos_a) is True or compatible_poisson(conteos_a) == True, "A es compatible con Poisson"
assert not compatible_poisson(conteos_b), "B no es compatible con Poisson"
print(f"razón A = {razon_dispersion(conteos_a):.3f}   razón B = {razon_dispersion(conteos_b):.3f}")
print("Ejercicio 4 correcto ✔")
```

### Solución del reto final

```python
def comparar_tlc(rng, reps=5000):
    """Devuelve {"exponencial": cociente, "cauchy": cociente} con cociente = IQR(n=1000) / IQR(n=10)."""
    generadores = {"exponencial": lambda size: rng.exponential(1.0, size),
                   "cauchy": lambda size: rng.standard_cauchy(size)}
    salida = {}
    for nombre, gen in generadores.items():
        iqr = {n: stats.iqr(gen((reps, n)).mean(axis=1)) for n in (10, 1000)}
        salida[nombre] = iqr[1000] / iqr[10]
    return salida

resultado = comparar_tlc(np.random.default_rng(2024))
conclusion = "La Cauchy no tiene media ni varianza finitas: la media de n datos tiene la misma distribución que un solo dato, así que el TLC (que exige varianza finita) no se aplica."

assert resultado is not None, "Completa la función comparar_tlc (TODO)"
assert 0.08 < resultado["exponencial"] < 0.125, "con varianza finita el rango debe reducirse ≈ 10 veces"
assert 0.8 < resultado["cauchy"] < 1.25, "en la Cauchy promediar no reduce el rango: cociente ≈ 1"
assert len(conclusion.strip()) > 20, "Escribe tu conclusión (TODO)"
print({k: round(v, 3) for k, v in resultado.items()})
print("Reto final correcto ✔")
```
