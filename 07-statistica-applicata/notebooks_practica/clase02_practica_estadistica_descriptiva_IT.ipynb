### Soluzione dell'esercizio 3

```python
def cinco_numeros(x):
    """Quartili, IQR e limiti della regola 1,5 · IQR di un campione."""
    q1, mediana, q3 = np.quantile(x, [0.25, 0.5, 0.75])
    iqr = q3 - q1
    return {"q1": q1, "mediana": mediana, "q3": q3, "iqr": iqr,
            "lim_inf": q1 - 1.5 * iqr, "lim_sup": q3 + 1.5 * iqr}

nueve_ej3 = pd.Series([1600, 1800, 1900, 2000, 2100, 2200, 2500, 2800, 9500])
res = cinco_numeros(nueve_ej3)
atipicos_ej3 = sorted(nueve_ej3[(nueve_ej3 < res["lim_inf"]) | (nueve_ej3 > res["lim_sup"])].tolist())

assert res is not None and atipicos_ej3 is not None, "Completa le righe contrassegnate con TODO"
assert np.isclose(res["q1"], 1900) and np.isclose(res["q3"], 2500), "Q1 e Q3: posizioni 0,25·(n−1) e 0,75·(n−1)"
assert np.isclose(res["mediana"], 2100), "la mediana è il quinto dato"
assert np.isclose(res["iqr"], 600), "IQR = Q3 − Q1"
assert np.isclose(res["lim_inf"], 1000) and np.isclose(res["lim_sup"], 3400), "limiti: Q1 − 1,5·IQR e Q3 + 1,5·IQR"
assert list(atipicos_ej3) == [9500], "quali valori restano fuori dai limiti?"
print(res)
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 4

```python
datos_z = np.concatenate([rng.normal(2000, 300, size=198), [9000, 15000]])

def zscores(x):
    """z-score di ogni valore: (x − media) / deviazione standard campionaria."""
    return (x - np.mean(x)) / np.std(x, ddof=1)

z_ej4 = zscores(datos_z)
assert z_ej4 is not None, "Completa le righe contrassegnate con TODO"

n_z = int((np.abs(z_ej4) > 3).sum())
q1_z, q3_z = np.quantile(datos_z, [0.25, 0.75])
n_iqr = int(((datos_z < q1_z - 1.5 * (q3_z - q1_z)) | (datos_z > q3_z + 1.5 * (q3_z - q1_z))).sum())

assert None not in (n_z, n_iqr), "Completa le righe contrassegnate con TODO"
assert abs(z_ej4.mean()) < 1e-9 and abs(z_ej4.std(ddof=1) - 1) < 1e-9, "uno z-score ha media 0 e deviazione standard (ddof=1) 1"
assert n_z == 2, "|z| > 3 dovrebbe segnalare i due valori inseriti"
assert n_iqr >= 2, "la regola dell'IQR deve segnalare anche i due valori inseriti"
print(f"|z| > 3 segnala {n_z} valori; la regola dell'IQR ne segnala {n_iqr}")
print("Esercizio 4 corretto ✔")
```

### Soluzione della sfida finale

```python
temperaturas = rng.uniform(40, 60, size=200)
sueldos_reto = np.concatenate([np.round(rng.lognormal(7.6, 0.35, size=200), 2), [9_000, 14_000, 250_000]])

def informe(x):
    """Rapporto descrittivo di un campione numerico."""
    s = pd.Series(x)
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    n_atipicos = int(((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum())
    asimetria = s.skew()
    return {"media": s.mean(), "mediana": s.median(), "desv_std": s.std(ddof=1), "iqr": iqr,
            "asimetria": asimetria, "n_atipicos_iqr": n_atipicos,
            "recomendacion": "mediana" if (n_atipicos > 0 or abs(asimetria) > 1) else "media"}

assert informe(temperaturas) is not None, "Completa le righe contrassegnate con TODO"
tabla = pd.DataFrame({"temperaturas": informe(temperaturas), "sueldos": informe(sueldos_reto)})
display(tabla)

assert informe(temperaturas)["recomendacion"] == "media", "temperature simmetriche e senza estremi: media o mediana?"
assert informe(sueldos_reto)["recomendacion"] == "mediana", "gli stipendi hanno estremi e asimmetria: media o mediana?"
assert informe(sueldos_reto)["n_atipicos_iqr"] >= 3, "i 3 valori inseriti devono essere segnalati come anomali"
assert informe(sueldos_reto)["media"] > informe(sueldos_reto)["mediana"], "con coda a destra, la media sta sopra la mediana"
assert abs(informe(temperaturas)["asimetria"]) < 0.5, "un insieme uniforme è approssimativamente simmetrico"
print("Sfida finale corretta ✔")
```
