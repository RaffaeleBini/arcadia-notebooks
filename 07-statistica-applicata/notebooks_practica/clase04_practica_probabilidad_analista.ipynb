### Solución del ejercicio 3

```python
def vpp(prevalencia, sensibilidad, especificidad):
    """P(enfermo | positivo) por el teorema de Bayes."""
    p_positivo = prevalencia * sensibilidad + (1 - prevalencia) * (1 - especificidad)
    return prevalencia * sensibilidad / p_positivo

assert vpp(0.01, 0.95, 0.95) is not None, "Completa la función vpp (TODO)"
assert np.isclose(vpp(0.01, 0.95, 0.95), 95 / 590), "caso guía: 95 verdaderos positivos entre 590 positivos"
assert np.isclose(vpp(0.02, 0.90, 0.97), 0.018 / 0.0474), "revisa la probabilidad total de dar positivo"
assert np.isclose(vpp(0.50, 0.95, 0.95), 0.95), "con prevalencia 50 % y sensibilidad = especificidad, el VPP es igual a ellas"
assert vpp(0.001, 0.95, 0.95) < vpp(0.01, 0.95, 0.95) < vpp(0.1, 0.95, 0.95), "el VPP debe crecer con la prevalencia"
print(f"VPP(1 %, 95 %, 95 %) = {vpp(0.01, 0.95, 0.95):.4f}   VPP(2 %, 90 %, 97 %) = {vpp(0.02, 0.90, 0.97):.4f}")
print("Ejercicio 3 correcto ✔")
```

### Solución del ejercicio 4

```python
def simular_vpp(prevalencia, sensibilidad, especificidad, n, rng):
    """Estimación por simulación de P(enfermo | positivo)."""
    enfermo = rng.random(n) < prevalencia
    u = rng.random(n)
    positivo = np.where(enfermo, u < sensibilidad, u < 1 - especificidad)
    return enfermo[positivo].mean()

rng_ej = np.random.default_rng(7)
estimacion = simular_vpp(0.05, 0.95, 0.95, 500_000, rng_ej)

assert estimacion is not None, "Completa la función simular_vpp (TODO)"
exacto = 0.05 * 0.95 / (0.05 * 0.95 + 0.95 * 0.05)          # = 0,5
n_pos_aprox = 500_000 * (0.05 * 0.95 + 0.95 * 0.05)
error_tipico = np.sqrt(exacto * (1 - exacto) / n_pos_aprox)
assert abs(estimacion - exacto) < 5 * error_tipico, "la estimación está demasiado lejos del valor exacto: ¿estás dividiendo entre los positivos?"
print(f"exacto = {exacto:.4f}   simulado = {estimacion:.4f}   (error típico ≈ {error_tipico:.4f})")
print("Ejercicio 4 correcto ✔")
```

### Solución del ejercicio 5

```python
tiradas = rng.integers(1, 7, size=600_000)
previa_es_6 = tiradas[:-1] == 6      # tirada anterior
actual_es_6 = tiradas[1:] == 6       # tirada siguiente

p_seis_tras_seis = actual_es_6[previa_es_6].mean()
p_dos_seises = (previa_es_6 & actual_es_6).mean()

assert None not in (p_seis_tras_seis, p_dos_seises), "Completa las líneas marcadas con TODO"
assert abs(p_seis_tras_seis - 1 / 6) < 0.01, "si los dados son independientes, debería estar cerca de 1/6"
assert abs(p_dos_seises - 1 / 36) < 0.002, "debería estar cerca de 1/36"
print(f"P(6 | anterior 6) ≈ {p_seis_tras_seis:.4f} (1/6 = {1/6:.4f})   P(6 y 6) ≈ {p_dos_seises:.4f} (1/36 = {1/36:.4f})")
print("Ejercicio 5 correcto ✔")
```

### Solución del reto final

```python
def informe(prevalencia, sensibilidad, especificidad, n=1_000_000):
    """Informe de un test: VPP y VPN exactos, VPP simulado y prevalencia mínima para VPP ≥ 50 %."""
    p_pos = prevalencia * sensibilidad + (1 - prevalencia) * (1 - especificidad)
    p_neg = 1 - p_pos
    vpp_e = prevalencia * sensibilidad / p_pos
    vpn_e = (1 - prevalencia) * especificidad / p_neg
    enfermo = rng.random(n) < prevalencia
    u = rng.random(n)
    positivo = np.where(enfermo, u < sensibilidad, u < 1 - especificidad)
    return {"vpp": vpp_e, "vpn": vpn_e, "vpp_sim": enfermo[positivo].mean(),
            "prev_50": (1 - especificidad) / (sensibilidad + 1 - especificidad)}

res = informe(0.005, 0.90, 0.98)

assert res is not None, "Completa la función informe (TODO)"
assert np.isclose(res["vpp"], 0.0045 / 0.0244), "revisa el VPP exacto"
assert np.isclose(res["vpn"], 0.9751 / 0.9756), "VPN = P(sano | negativo)"
assert np.isclose(res["prev_50"], 0.02 / 0.92), "prev = (1 − esp) / (sens + 1 − esp)"
assert abs(res["vpp_sim"] - res["vpp"]) < 0.015, "la simulación debería acercarse al VPP exacto"
print(pd.Series(res).round(4))
print("Reto final correcto ✔")
```
