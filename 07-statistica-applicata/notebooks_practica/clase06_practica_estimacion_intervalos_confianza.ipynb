### Solución del ejercicio 1

### Solución del ejercicio 2

```python
def semiancho(s, n, nivel):
    """Semiancho del IC para la media: t · s / √n."""
    t_crit = stats.t.ppf(1 - (1 - nivel) / 2, df=n - 1)
    return t_crit * s / np.sqrt(n)

sa_90 = semiancho(4.69, 9, 0.90)
sa_99 = semiancho(4.69, 9, 0.99)
factor_n4 = semiancho(4.69, 36, 0.95) / semiancho(4.69, 9, 0.95)

assert semiancho(4.69, 9, 0.95) is not None, "Completa la función semiancho (TODO)"
assert all(v is not None for v in (sa_90, sa_99, factor_n4)), "Completa las líneas marcadas con TODO"
assert np.isclose(semiancho(4.69, 9, 0.95), 2.306 * 4.69 / 3, rtol=1e-3), "t(0,975; 8) = 2,306"
assert sa_90 < semiancho(4.69, 9, 0.95) < sa_99, "más confianza, más ancho"
assert np.isclose(sa_99 / sa_90, 3.355 / 1.860, rtol=1e-2), "revisa los cuantiles t de 90 % y 99 %"
assert 0.43 < factor_n4 < 0.45, "con n cuatro veces mayor el cociente es un poco menor que 0,5 (por la t)"
print(f"sa_90 = {sa_90:.3f}   sa_99 = {sa_99:.3f}   factor_n4 = {factor_n4:.3f}")
print("Ejercicio 2 correcto ✔")
```

### Solución del ejercicio 3

```python
def cobertura(n, nivel, reps, rng, usar_t):
    """Proporción de IC que contienen la media verdadera (50)."""
    m = rng.normal(50, 10, size=(reps, n))
    ees = m.std(axis=1, ddof=1) / np.sqrt(n)
    q = 1 - (1 - nivel) / 2
    crit = stats.t.ppf(q, df=n - 1) if usar_t else stats.norm.ppf(q)
    medias = m.mean(axis=1)
    return ((medias - crit * ees <= 50) & (50 <= medias + crit * ees)).mean()

rng_ej = np.random.default_rng(123)
cob_t = cobertura(5, 0.95, 1_000, rng_ej, True)
cob_z = cobertura(5, 0.95, 1_000, rng_ej, False)

assert cob_t is not None, "Completa la función cobertura (TODO)"
assert 0.92 < cob_t < 0.98, "la t con n = 5 debería cubrir cerca de 0,95"
assert cob_z < 0.92, "con z y n = 5 la cobertura debería ser claramente menor (≈ 0,88)"
assert cob_z < cob_t - 0.03, "z debe cubrir menos que t"
print(f"cobertura con t = {cob_t:.3f}   cobertura con z = {cob_z:.3f}")
print("Ejercicio 3 correcto ✔")
```

### Solución del ejercicio 4

```python
datos = np.random.default_rng(11).lognormal(mean=np.log(30), sigma=0.5, size=30)     # 30 tiempos de entrega (ejemplo ilustrativo)

def bootstrap_ic(datos, estadistico, nivel, B, rng):
    """IC percentil bootstrap: devuelve (inferior, superior)."""
    idx = rng.integers(0, len(datos), size=(B, len(datos)))
    dist = estadistico(datos[idx], axis=1)
    alfa = 1 - nivel
    inf, sup = np.quantile(dist, [alfa / 2, 1 - alfa / 2])
    return inf, sup

rng_ej = np.random.default_rng(5)
ic_mediana = bootstrap_ic(datos, np.median, 0.95, 5_000, rng_ej)

assert ic_mediana is not None, "Completa la función bootstrap_ic (TODO)"
inferior, superior = ic_mediana
assert inferior < np.median(datos) < superior, "el IC debería contener la mediana de la muestra"
ref = np.quantile(np.median(datos[np.random.default_rng(999).integers(0, 30, size=(20_000, 30))], axis=1), [0.025, 0.975])
assert abs(inferior - ref[0]) < 1.5 and abs(superior - ref[1]) < 1.5, "tus límites se alejan de la referencia: ¿remuestreas con reposición y de tamaño n?"
print(f"Mediana muestral = {np.median(datos):.2f}   IC bootstrap 95 % = ({inferior:.2f}, {superior:.2f})")
print("Ejercicio 4 correcto ✔")
```

### Solución del reto final

```python
datos_reto = np.random.default_rng(2026).lognormal(mean=np.log(30), sigma=0.5, size=200)

def informe_ic(datos, nivel, B, rng):
    """IC de la media (t), de la media (bootstrap) y de la mediana (bootstrap)."""
    idx = rng.integers(0, len(datos), size=(B, len(datos)))
    remuestras = datos[idx]
    alfa = 1 - nivel
    cuantiles = [alfa / 2, 1 - alfa / 2]
    return {"n": len(datos),
            "t": stats.t.interval(nivel, df=len(datos) - 1, loc=datos.mean(), scale=stats.sem(datos)),
            "boot_media": tuple(np.quantile(remuestras.mean(axis=1), cuantiles)),
            "boot_mediana": tuple(np.quantile(np.median(remuestras, axis=1), cuantiles))}

res = informe_ic(datos_reto, 0.95, 10_000, np.random.default_rng(1))

assert res is not None, "Completa la función informe_ic (TODO)"
assert res["n"] == 200, "n es el tamaño de la muestra"
ref_t = stats.t.interval(0.95, df=199, loc=datos_reto.mean(), scale=stats.sem(datos_reto))
assert np.allclose(res["t"], ref_t), "revisa el IC con la t (n - 1 grados de libertad, error estándar s/√n)"
ancho_t = res["t"][1] - res["t"][0]
ancho_b = res["boot_media"][1] - res["boot_media"][0]
assert abs(ancho_b / ancho_t - 1) < 0.15, "con n = 200 el IC bootstrap de la media debe parecerse al de la t"
assert res["boot_media"][0] < datos_reto.mean() < res["boot_media"][1], "el IC bootstrap debe contener la media muestral"
assert res["boot_mediana"][0] < np.median(datos_reto) < res["boot_mediana"][1], "el IC de la mediana debe contener la mediana muestral"
print(pd.DataFrame({k: v for k, v in res.items() if k != "n"}, index=["inferior", "superior"]).round(2))
print("Reto final correcto ✔")
```
