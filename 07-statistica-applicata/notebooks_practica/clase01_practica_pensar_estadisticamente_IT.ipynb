### Soluzione dell'esercizio 1

```python
poblacion = rng.normal(50, 10, size=20_000)
muestra = rng.choice(poblacion, size=100, replace=False)

mu = poblacion.mean()
x_barra = muestra.mean()
error = x_barra - mu

assert np.isclose(mu, poblacion.mean()), "mu deve essere la media della POPOLAZIONE"
assert np.isclose(x_barra, muestra.mean()), "x_barra deve essere la media del CAMPIONE"
assert np.isclose(error, x_barra - mu), "error = x_barra - mu"
print(f"μ = {mu:.2f}   x̄ = {x_barra:.2f}   error = {error:+.2f}")
print("Esercizio 1 corretto ✔")
```

### Soluzione dell'esercizio 2

```python
from itertools import combinations

pob5 = [2, 4, 6, 8, 10]

medias = [np.mean(c) for c in combinations(pob5, 4)]
promedio_de_medias = np.mean(medias)

assert len(medias) == 5, "ci sono 5 campioni distinti di dimensione 4"
assert sorted(medias) == [5.0, 5.5, 6.0, 6.5, 7.0], "rivedi le medie di ogni campione"
assert np.isclose(promedio_de_medias, 6.0), "la media delle medie deve essere 6"
print("Medie:", sorted(medias), "→ media:", promedio_de_medias)
print("Esercizio 2 corretto ✔")
```

### Soluzione dell'esercizio 3

```python
poblacion_3 = rng.normal(100, 20, size=50_000)

def desv_medias(n, reps=2000):
    """Deviazione standard delle medie di `reps` campioni di dimensione `n`."""
    muestras = rng.choice(poblacion_3, size=(reps, n))
    return muestras.mean(axis=1).std()

d25, d100, d400 = desv_medias(25), desv_medias(100), desv_medias(400)
print(f"n=25: {d25:.2f}   n=100: {d100:.2f}   n=400: {d400:.2f}")

assert d25 > d100 > d400, "l'oscillazione deve diminuire all'aumentare di n"
assert abs(d100 - 20 / np.sqrt(100)) < 0.15 * 20 / np.sqrt(100), "dovrebbe assomigliare a σ/√n"
print("Esercizio 3 corretto ✔")
```

### Soluzione dell'esercizio 5

```python
N = 100_000
satisf = np.clip(rng.normal(6.0, 2.0, size=N), 0, 10)
resp = satisf[rng.random(N) < 1 / (1 + np.exp(-(satisf - 7)))]     # solo coloro che rispondono

error_aleatorio = abs(rng.choice(satisf, 500, replace=False).mean() - satisf.mean())
error_sesgado = abs(rng.choice(resp, 500, replace=False).mean() - satisf.mean())

assert error_aleatorio < 0.3, "il campione casuale dovrebbe stare vicino alla media reale"
assert error_sesgado > 0.3, "il campione distorto dovrebbe mostrare un errore chiaro"
print(f"error casuale = {error_aleatorio:.3f}   error distorto = {error_sesgado:.3f}")
print("Esercizio 5 corretto ✔")
```

### Soluzione della sfida finale

```python
def error_medio(n, sesgada, reps=200):
    """Errore assoluto medio della media campionaria dopo `reps` sondaggi di dimensione n."""
    fuente = resp if sesgada else satisf
    errores = [abs(rng.choice(fuente, n, replace=False).mean() - satisf.mean()) for _ in range(reps)]
    return float(np.mean(errores))

tabla = pd.DataFrame({
    "n": [100, 1_000, 10_000],
    "casuale": [error_medio(n, False) for n in [100, 1_000, 10_000]],
    "distorta": [error_medio(n, True) for n in [100, 1_000, 10_000]],
}).set_index("n").round(3)
display(tabla)

assert tabla["casuale"].is_monotonic_decreasing, "l'errore casuale dovrebbe scendere al crescere di n"
assert tabla.loc[10_000, "casuale"] < 0.05, "con n=10 000 il campione casuale dovrebbe essere molto preciso"
assert tabla.loc[10_000, "distorta"] > 0.3, "la distorsione dovrebbe persistere con n=10 000"
print("Sfida finale corretta ✔")
```
