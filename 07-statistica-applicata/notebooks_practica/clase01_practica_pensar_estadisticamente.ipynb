### Solución del ejercicio 5

```python
N = 100_000
satisf = np.clip(rng.normal(6.0, 2.0, size=N), 0, 10)
resp = satisf[rng.random(N) < 1 / (1 + np.exp(-(satisf - 7)))]     # solo quienes responden

error_aleatorio = abs(rng.choice(satisf, 500, replace=False).mean() - satisf.mean())
error_sesgado = abs(rng.choice(resp, 500, replace=False).mean() - satisf.mean())

assert error_aleatorio < 0.3, "la muestra aleatoria debería estar cerca de la media real"
assert error_sesgado > 0.3, "la muestra sesgada debería mostrar un error claro"
print(f"error aleatorio = {error_aleatorio:.3f}   error sesgado = {error_sesgado:.3f}")
print("Ejercicio 5 correcto ✔")
```

### Solución del reto final

```python
def error_medio(n, sesgada, reps=200):
    """Error absoluto medio de la media muestral tras `reps` encuestas de tamaño n."""
    fuente = resp if sesgada else satisf
    errores = [abs(rng.choice(fuente, n, replace=False).mean() - satisf.mean()) for _ in range(reps)]
    return float(np.mean(errores))

tabla = pd.DataFrame({
    "n": [100, 1_000, 10_000],
    "aleatoria": [error_medio(n, False) for n in [100, 1_000, 10_000]],
    "sesgada": [error_medio(n, True) for n in [100, 1_000, 10_000]],
}).set_index("n").round(3)
display(tabla)

assert tabla["aleatoria"].is_monotonic_decreasing, "el error aleatorio debería bajar al crecer n"
assert tabla.loc[10_000, "aleatoria"] < 0.05, "con n=10 000 la muestra aleatoria debería ser muy precisa"
assert tabla.loc[10_000, "sesgada"] > 0.3, "el sesgo debería persistir con n=10 000"
print("Reto final correcto ✔")
```
