### Soluzione dell'esercizio 4

```python
def razon_dispersion(x):
    """Varianza campionaria (ddof=1) divisa per la media."""
    return np.var(x, ddof=1) / np.mean(x)

def compatible_poisson(x, tol=0.25):
    """True se il rapporto varianza/media dista meno di tol da 1."""
    return bool(abs(razon_dispersion(x) - 1) < tol)

rng_ej = np.random.default_rng(5)
conteos_a = rng_ej.poisson(5, size=50_000)
conteos_b = rng_ej.negative_binomial(n=2.5, p=0.25, size=50_000)      # media 7,5 e varianza 30

assert razon_dispersion(np.array([1.0, 2.0, 3.0])) is not None, "Completa le funzioni (TODO)"
assert np.isclose(razon_dispersion(np.array([1.0, 2.0, 3.0])), 0.5), "varianza(ddof=1) = 1 e media = 2 → rapporto 0,5"
assert 0.95 < razon_dispersion(conteos_a) < 1.05, "i dati Poisson devono dare un rapporto vicino a 1"
assert 3.5 < razon_dispersion(conteos_b) < 4.5, "i dati sovradispersi hanno varianza / media = 4"
assert compatible_poisson(conteos_a) is True or compatible_poisson(conteos_a) == True, "A è compatibile con la Poisson"
assert not compatible_poisson(conteos_b), "B non è compatibile con la Poisson"
print(f"rapporto A = {razon_dispersion(conteos_a):.3f}   rapporto B = {razon_dispersion(conteos_b):.3f}")
print("Esercizio 4 corretto ✔")
```

### Soluzione della sfida finale

```python
def comparar_tlc(rng, reps=5000):
    """Restituisce {"exponencial": rapporto, "cauchy": rapporto} con rapporto = IQR(n=1000) / IQR(n=10)."""
    generadores = {"exponencial": lambda size: rng.exponential(1.0, size),
                   "cauchy": lambda size: rng.standard_cauchy(size)}
    salida = {}
    for nombre, gen in generadores.items():
        iqr = {n: stats.iqr(gen((reps, n)).mean(axis=1)) for n in (10, 1000)}
        salida[nombre] = iqr[1000] / iqr[10]
    return salida

resultado = comparar_tlc(np.random.default_rng(2024))
conclusion = "La Cauchy non ha media né varianza finite: la media di n dati ha la stessa distribuzione di un solo dato, quindi il TLC (che richiede varianza finita) non si applica."

assert resultado is not None, "Completa la funzione comparar_tlc (TODO)"
assert 0.08 < resultado["exponencial"] < 0.125, "con varianza finita lo scarto deve ridursi di ≈ 10 volte"
assert 0.8 < resultado["cauchy"] < 1.25, "nella Cauchy fare la media non riduce lo scarto: rapporto ≈ 1"
assert len(conclusion.strip()) > 20, "Scrivi la tua conclusione (TODO)"
print({k: round(v, 3) for k, v in resultado.items()})
print("Sfida finale corretta ✔")
```
