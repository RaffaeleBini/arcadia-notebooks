## Cella 2 — Prima installazione e chiamata minima

Segui il flusso descritto nella dispensa: installare la libreria (già fatto
nei prerequisiti), importarla, creare un client, inviare il prompt, leggere
la risposta. Qui **non** costruiamo ancora una funzione wrapper — lo faremo
nel Modulo 08. Ci limitiamo alla chiamata più semplice possibile.

````python
from openai import OpenAI

client = OpenAI(api_key=OPENAI_KEY)

risposta = client.chat.completions.create(
    model="gpt-5-mini",
    max_tokens=100,
    messages=[
        {"role": "user", "content": "In una frase: cos'è un'API?"}
    ]
)

print(risposta.choices[0].message.content)
````

---
## Cella 3 — Osserva la struttura completa della risposta

La risposta dell'API non è solo testo: è un oggetto strutturato con molte
informazioni utili. Esploriamola prima di passare, nel Modulo 08, a
incapsularla dentro funzioni riutilizzabili.

````python
print("ID della risposta:      ", risposta.id)
print("Modello effettivamente usato:", risposta.model)
print("Motivo di terminazione:", risposta.choices[0].finish_reason)
print()
print("Token di input:  ", risposta.usage.prompt_tokens)
print("Token di output: ", risposta.usage.completion_tokens)
print("Token totali:    ", risposta.usage.total_tokens)
````
