import json

def cargar_historial():
    """Carga el historial de conversacion guardado, si existe."""
    if os.path.exists(ARCHIVO_HISTORIAL):
        with open(ARCHIVO_HISTORIAL, "r", encoding="utf-8") as f:

## Reto 3 — Añade una tercera clase de daño al clasificador de imágenes (abierto)

**Caso de uso base**: el clasificador de la clase 12 distingue entre dos categorías de daño en aeronaves: abolladura (*dent*) y grieta (*crack*), usando transfer learning con VGG16.

**Reto**: extiende el clasificador para que reconozca una **tercera** categoría —por ejemplo, "corrosión" (*corrosion*) o "sin daño" (*no damage*)— convirtiendo el problema de clasificación binaria en clasificación multiclase.

**Pistas para empezar**:
- La capa de salida del modelo tendría que cambiar de `Dense(1, activation='sigmoid')` (una sola probabilidad, para 2 clases) a `Dense(3, activation='softmax')` (una probabilidad por cada una de las 3 clases).
- La función de pérdida cambiaría de `binary_crossentropy` a `categorical_crossentropy` (o `sparse_categorical_crossentropy`, si no haces one-hot encoding de las etiquetas).
- Necesitarías organizar tus imágenes en tres subcarpetas (`train/dent`, `train/crack`, `train/corrosion`, por ejemplo) para que `flow_from_directory` con `class_mode='categorical'` las reconozca automáticamente.
- Reflexiona: ¿cuántas imágenes de la nueva categoría necesitarías como mínimo para que el modelo aprenda algo razonable? ¿Qué pasaría si tuvieras muchas menos fotos de "corrosión" que de las otras dos clases (un dataset desbalanceado)?
