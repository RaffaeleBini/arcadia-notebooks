```python
X_and = np.array([[0, 0, 1, 1],
                   [0, 1, 0, 1]])
d_and = np.array([[0, 0, 0, 1]])

inputSize = 2
hiddenSize = 2
outputSize = 1
lr = 0.5
epochs = 10000

w1 = np.random.uniform(size=(hiddenSize, inputSize), low=-1, high=1)
b1 = np.random.uniform(size=(hiddenSize, 1), low=-1, high=1)
w2 = np.random.uniform(size=(outputSize, hiddenSize), low=-1, high=1)
b2 = np.random.uniform(size=(outputSize, 1), low=-1, high=1)

error_list = []

for epoch in range(epochs):
    # forward pass
    z1 = np.dot(w1, X_and) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(w2, a1) + b2
    a2 = sigmoid(z2)

    # errore
    error = d_and - a2

    # backward pass
    dz2 = error * a2 * (1 - a2)
    dw2 = np.dot(dz2, a1.T)
    db2 = np.sum(dz2, axis=1, keepdims=True)

    da1 = np.dot(w2.T, dz2)
    dz1 = da1 * a1 * (1 - a1)
    dw1 = np.dot(dz1, X_and.T)
    db1 = np.sum(dz1, axis=1, keepdims=True)

    # aggiornamento
    w2 += lr * dw2
    b2 += lr * db2
    w1 += lr * dw1
    b1 += lr * db1

    error_list.append(np.mean(np.abs(error)))

    if epoch % 1000 == 0:
        print('Epoca {}: errore medio = {:.5f}'.format(epoch, np.mean(np.abs(error))))

# test finale
z1 = np.dot(w1, X_and) + b1
a1 = sigmoid(z1)
z2 = np.dot(w2, a1) + b2
a2 = sigmoid(z2)

print('Previsioni finali:', np.round(a2, 5))
print('Valori attesi:     ', d_and)

plt.plot(error_list)
plt.title('Errore medio lungo le epoche (AND)')
plt.xlabel('Epoca')
plt.ylabel('Errore medio assoluto')
plt.grid(True)
plt.show()
```

Noterai che l'errore di AND scende molto più velocemente di quello di XOR con lo stesso tasso di
