```python
import cirq

# Dos qubits físicos en la rejilla
qubit1 = cirq.GridQubit(0, 0)
qubit2 = cirq.GridQubit(0, 1)

# El circuito acumula las operaciones que aplicas a los qubits
circuito = cirq.Circuit()

# La puerta Hadamard pone el primer qubit en superposición
circuito.append(cirq.H(qubit1))

# La puerta CNOT entrelaza el segundo qubit con el primero
circuito.append(cirq.CNOT(qubit1, qubit2))

# Medimos los dos qubits juntos, bajo la clave "resultado"
circuito.append(cirq.measure(qubit1, qubit2, key="resultado"))

print(circuito)

# Simulamos el circuito 1000 veces
simulador = cirq.Simulator()
resultados = simulador.run(circuito, repetitions=1000)

# Solo deberían aparecer las combinaciones "00" y "11"
print(resultados.histogram(key="resultado"))
```
