```python
import cirq

# Tre qubit nella griglia, uno per riga
q0, q1, q2 = cirq.GridQubit(0, 0), cirq.GridQubit(1, 0), cirq.GridQubit(2, 0)

circuito = cirq.Circuit()

# Portiamo il primo qubit in sovrapposizione
circuito.append(cirq.H(q0))

# Colleghiamo q0 con q1 in entanglement
circuito.append(cirq.CNOT(q0, q1))

# Colleghiamo q1 con q2: l'entanglement si propaga a catena
circuito.append(cirq.CNOT(q1, q2))

# Misuriamo i tre qubit insieme
circuito.append(cirq.measure(q0, q1, q2, key="resultado"))

print(circuito)

# Simuliamo 1.000 ripetizioni
simulador = cirq.Simulator()
resultado = simulador.run(circuito, repetitions=1000)

# L'istogramma deve mostrare solo due risultati possibili: "000" e "111"
# (in formato intero: 0 e 7), perché i tre qubit restano perfettamente
# correlati grazie all'entanglement a catena. Questo è lo stato GHZ.
print(resultado.histogram(key="resultado"))
```
