```python
import cirq

# Tres qubits en la rejilla, uno por fila
q0, q1, q2 = cirq.GridQubit(0, 0), cirq.GridQubit(1, 0), cirq.GridQubit(2, 0)

circuito = cirq.Circuit()

# Ponemos el primer qubit en superposición
circuito.append(cirq.H(q0))

# Entrelazamos q0 con q1
circuito.append(cirq.CNOT(q0, q1))

# Entrelazamos q1 con q2: el entrelazamiento se propaga en cadena
circuito.append(cirq.CNOT(q1, q2))

# Medimos los tres qubits a la vez
circuito.append(cirq.measure(q0, q1, q2, key="resultado"))

print(circuito)

# Simulamos 1.000 repeticiones
simulador = cirq.Simulator()
resultado = simulador.run(circuito, repetitions=1000)

# El histograma solo debe mostrar dos resultados posibles: "000" y "111"
# (en formato entero: 0 y 7), porque los tres qubits quedan perfectamente
# correlacionados gracias al entrelazamiento en cadena. Este es el estado GHZ.
print(resultado.histogram(key="resultado"))
```
