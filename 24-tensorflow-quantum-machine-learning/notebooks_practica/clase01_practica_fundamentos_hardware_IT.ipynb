```python
import cirq

# Due qubit fisici nella griglia
qubit1 = cirq.GridQubit(0, 0)
qubit2 = cirq.GridQubit(0, 1)

# Il circuito accumula le operazioni che applichi ai qubit
circuito = cirq.Circuit()

# La porta di Hadamard porta il primo qubit in sovrapposizione
circuito.append(cirq.H(qubit1))

# La porta CNOT crea l'entanglement tra il secondo qubit e il primo
circuito.append(cirq.CNOT(qubit1, qubit2))

# Misuriamo i due qubit insieme, sotto la chiave "resultado"
circuito.append(cirq.measure(qubit1, qubit2, key="resultado"))

print(circuito)

# Simuliamo il circuito 1000 volte
simulador = cirq.Simulator()
resultados = simulador.run(circuito, repetitions=1000)

# Devono comparire solo le combinazioni "00" e "11"
print(resultados.histogram(key="resultado"))
```
