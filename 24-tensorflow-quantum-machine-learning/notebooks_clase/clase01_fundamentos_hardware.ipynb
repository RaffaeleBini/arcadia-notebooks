## El bit clásico frente al qubit

Un bit clásico es como un interruptor: está en 0 o está en 1, nunca en ambos a la vez.

Un qubit (bit cuántico) es como una moneda que gira en el aire: mientras gira, no es ni cara ni cruz, es una combinación de las dos. Solo cuando la moneda cae —cuando mides el qubit— se decide si el resultado es 0 o 1.

A esa combinación la llamas **superposición**. Matemáticamente, describes un qubit como una mezcla ponderada de los estados 0 y 1, y esos pesos determinan la probabilidad de obtener cada resultado al medir.

Cuantos más qubits combinas en superposición, más combinaciones representas a la vez: con solo diez qubits ya representas 1.024 combinaciones simultáneas, algo que a un ordenador clásico le costaría 1.024 pasos independientes.

## Puertas cuánticas: cómo manipulas un qubit

Un ordenador clásico usa puertas lógicas (AND, OR, NOT) para transformar bits. Un ordenador cuántico usa **puertas cuánticas** para transformar qubits. La diferencia clave: las puertas cuánticas son reversibles, siempre puedes deshacer la operación.

Las que más vas a encontrar en este curso son:

- **Puerta Hadamard (H)**: coloca un qubit en superposición perfecta (50% de probabilidad de 0, 50% de probabilidad de 1)
- **Puertas X, Y, Z**: rotan el estado del qubit sobre los tres ejes de una esfera imaginaria (la esfera de Bloch)
- **Puerta CNOT**: actúa sobre dos qubits a la vez y es la base del entrelazamiento

## Entrelazamiento: cuando dos qubits dejan de ser independientes

Cuando entrelazas dos qubits, el estado de uno queda ligado al estado del otro, sin importar la distancia física entre ellos. Si mides el primer qubit y obtienes un 0, sabes automáticamente qué valor va a salir en el segundo, aunque todavía no lo hayas medido.

Einstein llamó a esto "acción fantasmal a distancia" porque le resultaba perturbador, pero hoy es una herramienta de trabajo diaria en los laboratorios de computación cuántica. El entrelazamiento es el ingrediente que permite que los qubits colaboren entre sí para resolver un problema, en lugar de comportarse como bits independientes.

## Medir: el paso que lo cambia todo

Mientras no mides un qubit, vive en superposición. En el instante en que lo mides, ese qubit colapsa a un valor concreto (0 o 1) y pierde toda la información de la superposición anterior.

Este comportamiento no tiene equivalente en la computación clásica y explica por qué programar en cuántica exige pensar de forma distinta: diseñas un circuito para que, al medir al final, la probabilidad de obtener la respuesta correcta sea alta.

## Manos a la obra: tu primer circuito con Cirq

Vas a usar Cirq, la librería de Google para programar circuitos cuánticos (la vas a usar en detalle en la próxima clase). Aquí tienes un primer vistazo: un qubit que pones en superposición con una puerta Hadamard y después mides.

```python
import cirq

# Un qubit físico situado en la posición (0, 0) de una rejilla
qubit = cirq.GridQubit(0, 0)

# Un circuito es la secuencia de operaciones que aplicas al qubit
circuito = cirq.Circuit()

# La puerta Hadamard pone el qubit en superposición: 50% 0, 50% 1
circuito.append(cirq.H(qubit))

# Medimos el qubit y guardamos el resultado con la clave "resultado"
circuito.append(cirq.measure(qubit, key="resultado"))

print(circuito)

# Simulamos el circuito 1000 veces para ver la distribución de resultados
simulador = cirq.Simulator()
resultados = simulador.run(circuito, repetitions=1000)

print(resultados.histogram(key="resultado"))
# Debería salir aproximadamente 500 veces "0" y 500 veces "1"
```

Fíjate en algo importante: cada vez que ejecutas el circuito completo (superposición y medida), obtienes un único 0 o un único 1, nunca los dos a la vez. Por eso repites el experimento 1.000 veces: así compruebas que, en conjunto, la moneda cayó cara la mitad de las veces y cruz la otra mitad, tal y como predice la superposición.

## Ahora con entrelazamiento: dos qubits que dejan de ser independientes

Amplías el circuito anterior: en vez de un único qubit, usas dos. Aplicas una puerta Hadamard al primero y después una puerta `cirq.CNOT(qubit1, qubit2)` para entrelazarlos. Mides ambos qubits y ejecutas el circuito 1.000 veces.

```python
# Dos qubits físicos en la rejilla
qubit1 = cirq.GridQubit(0, 0)
qubit2 = cirq.GridQubit(0, 1)

circuito_entrelazado = cirq.Circuit()

# Hadamard pone el primer qubit en superposición
circuito_entrelazado.append(cirq.H(qubit1))

# CNOT entrelaza el segundo qubit con el primero
circuito_entrelazado.append(cirq.CNOT(qubit1, qubit2))

# Medimos los dos qubits juntos, con la clave "resultado"
circuito_entrelazado.append(cirq.measure(qubit1, qubit2, key="resultado"))

print(circuito_entrelazado)

simulador = cirq.Simulator()
resultados = simulador.run(circuito_entrelazado, repetitions=1000)

print(resultados.histogram(key="resultado"))
# Solo deberían aparecer las combinaciones "00" y "11", nunca "01" ni "10"
```

Observa el histograma: solo aparecen las combinaciones "00" y "11", nunca "01" ni "10". Eso es el entrelazamiento en acción, los dos qubits siempre coinciden.

## De la teoría al chip: por qué importa el hardware

Hasta ahora has hablado de qubits como si fueran monedas girando en el aire. Pero un qubit real es un objeto físico, y construirlo es una de las hazañas de ingeniería más difíciles de la actualidad.

Entender cómo se fabrica un qubit te ayuda a comprender por qué la computación cuántica actual tiene límites (ruido, pocos qubits, tiempos de coherencia cortos) y por qué muchos algoritmos de este curso están diseñados precisamente para convivir con esas limitaciones.

A esta etapa de la tecnología la conoces como la era **NISQ** (cuántico de escala intermedia y ruidoso): tienes ordenadores cuánticos, pero todavía imperfectos.

## El qubit superconductor: un circuito que se comporta como un átomo

La forma más habitual de construir un qubit hoy en día (la que usan Google o IBM) es el **qubit superconductor**, también llamado transmón. En vez de usar un átomo real, fabricas un circuito eléctrico diminuto que, enfriado a temperaturas extremas, se comporta igual que un átomo con solo dos niveles de energía posibles: el 0 y el 1.

La pieza clave de ese circuito es la **unión Josephson**: dos trozos de metal superconductor separados por una capa aislante ultrafina. Es como una compuerta que deja pasar corriente eléctrica de una forma muy particular, sin resistencia y con un comportamiento cuántico. Gracias a esa unión, el circuito completo se comporta como si tuviera niveles de energía discretos, exactamente igual que un átomo.

## Por qué hace tanto frío dentro de un ordenador cuántico

Para que el circuito se comporte como un qubit y no como un cable normal, el metal tiene que estar en estado **superconductor** (sin resistencia eléctrica), y eso solo ocurre a temperaturas cercanas al cero absoluto.

Por eso los ordenadores cuánticos de Google viven dentro de un **refrigerador de dilución**: una estructura en forma de araña metálica que enfría el chip hasta unos 15 milikelvin, más frío que el espacio exterior.

## Cómo controlas un qubit sin destruirlo

Controlar un qubit significa mandarle pulsos de microondas muy precisos para aplicarle las puertas cuánticas que viste antes. El problema: un pulso mal diseñado puede excitar el qubit a un nivel de energía que no quieres (un "estado fantasma"), estropeando el cálculo.

La técnica **DRAG** (eliminación derivativa por puerta adiabática) da forma a esos pulsos de microondas para evitar justo ese error, y es una de las piezas que hace posible controlar los qubits con la precisión necesaria.

Además, toda esa electrónica de control tiene que vivir muy cerca del chip cuántico para no perder precisión por el camino. Por eso también se investiga el **CMOS criogénico** (semiconductor complementario de óxido metálico): chips de control clásicos diseñados para funcionar a esas mismas temperaturas extremas, en vez de a temperatura ambiente.

## El enemigo común: la decoherencia

Por mucho cuidado que pongas, un qubit real siempre está expuesto a vibraciones, campos magnéticos residuales o imperfecciones del material, y con el tiempo pierde su estado cuántico: eso se llama **decoherencia**.

El tiempo que un qubit aguanta en superposición antes de decoherer se llama **tiempo de coherencia**, y cuanto más corto es, menos operaciones puedes encadenar antes de que el resultado deje de ser fiable.

Esta es la razón principal por la que los algoritmos avanzados que vas a ver más adelante en el curso están diseñados como circuitos cortos y poco profundos: cuanto menos dure el circuito, menos tiempo tiene la decoherencia para arruinar el resultado.

## Resumen de puntos clave

- Un qubit puede estar en superposición: una combinación de 0 y 1 al mismo tiempo, hasta que lo mides
- Las puertas cuánticas transforman qubits de forma reversible; la Hadamard crea superposición y la CNOT ayuda a crear entrelazamiento
- El entrelazamiento liga el estado de varios qubits entre sí, incluso a distancia
- Medir un qubit colapsa su superposición a un único valor concreto (0 o 1)
- Un qubit superconductor (transmón) es un circuito eléctrico que, enfriado casi al cero absoluto, se comporta como un átomo con dos niveles de energía
- La unión Josephson hace posible ese comportamiento, y el refrigerador de dilución mantiene el chip a temperaturas de milikelvin
- La técnica DRAG da forma a los pulsos de control y el CMOS criogénico acerca la electrónica al chip
- La decoherencia es la principal limitación de la era NISQ y explica por qué los circuitos actuales deben ser cortos
