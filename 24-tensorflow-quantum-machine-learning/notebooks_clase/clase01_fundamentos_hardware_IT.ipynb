## Il bit classico contro il qubit

Un bit classico è come un interruttore: è su 0 oppure su 1, mai su entrambi allo stesso tempo.

Un qubit (bit quantistico) è come una moneta che gira in aria: mentre gira, non è né testa né croce, è una combinazione dei due. Solo quando la moneta cade, cioè quando misuri il qubit, si decide se il risultato è 0 o 1.

Questa combinazione si chiama **sovrapposizione**. Matematicamente, descrivi un qubit come una miscela pesata degli stati 0 e 1, e questi pesi determinano la probabilità di ottenere ciascun risultato alla misura.

Più qubit combini in sovrapposizione, più combinazioni rappresenti contemporaneamente: con solo dieci qubit rappresenti già 1.024 combinazioni simultanee, un lavoro che per un computer classico richiederebbe 1.024 passi indipendenti.

## Porte quantistiche: come manipoli un qubit

Un computer classico usa porte logiche (AND, OR, NOT) per trasformare i bit. Un computer quantistico usa **porte quantistiche** per trasformare i qubit. La differenza chiave: le porte quantistiche sono reversibili, puoi sempre annullare l'operazione.

Quelle che incontrerai più spesso in questo corso sono:

- **Porta di Hadamard (H)**: porta un qubit in una sovrapposizione perfetta (probabilità del 50% di 0 e del 50% di 1)
- **Porte X, Y, Z**: ruotano lo stato del qubit attorno ai tre assi di una sfera immaginaria (la sfera di Bloch)
- **Porta CNOT**: agisce su due qubit contemporaneamente ed è la base dell'entanglement

## Entanglement: quando due qubit smettono di essere indipendenti

Quando crei entanglement tra due qubit, lo stato dell'uno resta legato allo stato dell'altro, indipendentemente dalla distanza fisica che li separa. Se misuri il primo qubit e ottieni 0, sai automaticamente quale valore uscirà per il secondo, anche se non lo hai ancora misurato.

Einstein chiamò questo fenomeno "azione spettrale a distanza" perché lo trovava inquietante, ma oggi è uno strumento di lavoro quotidiano nei laboratori di calcolo quantistico. L'entanglement è l'ingrediente che permette ai qubit di collaborare per risolvere un problema, invece di comportarsi come bit indipendenti.

## Misurare: il passo che cambia tutto

Finché non misuri un qubit, questo vive in sovrapposizione. Nell'istante in cui lo misuri, il qubit collassa su un valore preciso (0 o 1) e perde tutte le informazioni della sovrapposizione precedente.

Questo comportamento non ha equivalenti nel calcolo classico e spiega perché programmare in ambito quantistico richiede di pensare in modo diverso: progetti un circuito in modo che, misurando alla fine, la probabilità di ottenere la risposta corretta sia alta.

## Mettiamoci al lavoro: il tuo primo circuito con Cirq

Userai Cirq, la libreria di Google per programmare circuiti quantistici (la vedrai in dettaglio nella prossima lezione). Ecco un primo assaggio: un qubit che porti in sovrapposizione con una porta di Hadamard e poi misuri.

```python
import cirq

# Un qubit físico situado en la posición (0, 0) de una rejilla
qubit = cirq.GridQubit(0, 0)

# Un circuito es la secuencia de operaciones que aplicas al qubit
circuito = cirq.Circuit()

# La puerta Hadamard pone el qubit en superposición: 50% 0, 50% 1
circuito.append(cirq.H(qubit))

# Medimos el qubit y guardamos el resultado con la clave "resultado"
circuito.append(cirq.measure(qubit, key="resultado"))

print(circuito)

# Simulamos el circuito 1000 veces para ver la distribución de resultados
simulador = cirq.Simulator()
resultados = simulador.run(circuito, repetitions=1000)

print(resultados.histogram(key="resultado"))
# Debería salir aproximadamente 500 veces "0" y 500 veces "1"
```

Fai attenzione a una cosa importante: ogni volta che esegui il circuito completo (sovrapposizione e misura), ottieni un solo 0 o un solo 1, mai entrambi insieme. Per questo ripeti l'esperimento 1.000 volte: così verifichi che, nel complesso, la moneta è uscita testa circa la metà delle volte e croce l'altra metà, come prevede la sovrapposizione.

## Ora con l'entanglement: due qubit che smettono di essere indipendenti

Ampli il circuito precedente: invece di un solo qubit, ne usi due. Applichi una porta di Hadamard al primo e poi una porta `cirq.CNOT(qubit1, qubit2)` per creare l'entanglement tra i due. Misuri entrambi i qubit ed esegui il circuito 1.000 volte.

```python
# Dos qubits físicos en la rejilla
qubit1 = cirq.GridQubit(0, 0)
qubit2 = cirq.GridQubit(0, 1)

circuito_entrelazado = cirq.Circuit()

# Hadamard pone el primer qubit en superposición
circuito_entrelazado.append(cirq.H(qubit1))

# CNOT entrelaza el segundo qubit con el primero
circuito_entrelazado.append(cirq.CNOT(qubit1, qubit2))

# Medimos los dos qubits juntos, con la clave "resultado"
circuito_entrelazado.append(cirq.measure(qubit1, qubit2, key="resultado"))

print(circuito_entrelazado)

simulador = cirq.Simulator()
resultados = simulador.run(circuito_entrelazado, repetitions=1000)

print(resultados.histogram(key="resultado"))
# Solo deberían aparecer las combinaciones "00" y "11", nunca "01" ni "10"
```

Osserva l'istogramma: compaiono solo le combinazioni "00" e "11", mai "01" né "10". Questo è l'entanglement in azione: i due qubit coincidono sempre.

## Dalla teoria al chip: perché conta l'hardware

Finora hai parlato dei qubit come di monete che girano in aria. Ma un qubit reale è un oggetto fisico, e costruirlo è una delle imprese di ingegneria più difficili del momento.

Capire come si fabbrica un qubit ti aiuta a comprendere perché il calcolo quantistico attuale ha dei limiti (rumore, pochi qubit, tempi di coerenza brevi) e perché molti algoritmi di questo corso sono progettati proprio per convivere con quelle limitazioni.

Questa fase della tecnologia è nota come era **NISQ** (quantistico a scala intermedia e rumoroso): esistono computer quantistici, ma sono ancora imperfetti.

## Il qubit superconduttivo: un circuito che si comporta come un atomo

Il modo più diffuso oggi per costruire un qubit (quello usato da Google o IBM) è il **qubit superconduttivo**, detto anche transmon. Invece di usare un atomo reale, si fabbrica un minuscolo circuito elettrico che, raffreddato a temperature estreme, si comporta come un atomo con soli due livelli di energia possibili: lo 0 e l'1.

Il componente chiave di quel circuito è la **giunzione Josephson**: due pezzi di metallo superconduttore separati da uno strato isolante ultrasottile. È come una porta che lascia passare la corrente elettrica in modo molto particolare, senza resistenza e con un comportamento quantistico. Grazie a quella giunzione, l'intero circuito si comporta come se avesse livelli di energia discreti, esattamente come un atomo.

## Perché dentro un computer quantistico fa così freddo

Perché il circuito si comporti come un qubit e non come un normale cavo, il metallo deve trovarsi nello stato **superconduttivo** (senza resistenza elettrica), e questo accade solo a temperature vicine allo zero assoluto.

Per questo i computer quantistici di Google vivono dentro un **refrigeratore a diluizione**: una struttura metallica a forma di ragno che raffredda il chip fino a circa 15 millikelvin, più freddo dello spazio esterno.

## Come controlli un qubit senza distruggerlo

Controllare un qubit significa inviargli impulsi a microonde molto precisi per applicare le porte quantistiche che hai visto prima. Il problema: un impulso mal progettato può eccitare il qubit a un livello di energia che non vuoi (uno "stato fantasma"), rovinando il calcolo.

La tecnica **DRAG** (eliminazione derivativa mediante porta adiabatica) modella quegli impulsi a microonde per evitare proprio questo errore, ed è uno dei tasselli che rendono possibile controllare i qubit con la precisione necessaria.

Inoltre, tutta l'elettronica di controllo deve trovarsi molto vicina al chip quantistico per non perdere precisione lungo il percorso. Per questo si studia anche il **CMOS criogenico** (semiconduttore complementare a ossido metallico): chip di controllo classici progettati per funzionare alle stesse temperature estreme, invece che a temperatura ambiente.

## Il nemico comune: la decoerenza

Per quanta cura tu ci metta, un qubit reale è sempre esposto a vibrazioni, campi magnetici residui o imperfezioni del materiale, e nel tempo perde il suo stato quantistico: questo si chiama **decoerenza**.

Il tempo che un qubit resiste in sovrapposizione prima di decoerere si chiama **tempo di coerenza**, e più è breve, meno operazioni puoi concatenare prima che il risultato smetta di essere affidabile.

Questa è la ragione principale per cui gli algoritmi avanzati che vedrai più avanti nel corso sono progettati come circuiti brevi e poco profondi: più breve è il circuito, meno tempo ha la decoerenza per rovinare il risultato.

## Riepilogo dei punti chiave

- Un qubit può trovarsi in sovrapposizione: una combinazione di 0 e 1 allo stesso tempo, finché non lo misuri
- Le porte quantistiche trasformano i qubit in modo reversibile; la Hadamard crea sovrapposizione e la CNOT aiuta a creare entanglement
- L'entanglement lega tra loro lo stato di più qubit, anche a distanza
- Misurare un qubit fa collassare la sua sovrapposizione su un unico valore preciso (0 o 1)
- Un qubit superconduttivo (transmon) è un circuito elettrico che, raffreddato quasi allo zero assoluto, si comporta come un atomo con due livelli di energia
- La giunzione Josephson rende possibile questo comportamento, e il refrigeratore a diluizione mantiene il chip a temperature di millikelvin
- La tecnica DRAG modella gli impulsi di controllo e il CMOS criogenico avvicina l'elettronica al chip
- La decoerenza è la principale limitazione dell'era NISQ e spiega perché i circuiti attuali devono essere brevi
