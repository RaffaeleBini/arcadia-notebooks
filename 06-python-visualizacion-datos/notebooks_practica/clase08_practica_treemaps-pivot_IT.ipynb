```python
# Dati di esempio: spese personali per mese e categoria
data_mes = {
    'Mese': ['Gennaio', 'Gennaio', 'Gennaio',
            'Febbraio', 'Febbraio', 'Febbraio',
            'Marzo', 'Marzo', 'Marzo'],
    'Categoria': ['Cibo', 'Trasporto', 'Tempo libero',
                  'Cibo', 'Trasporto', 'Tempo libero',
                  'Cibo', 'Trasporto', 'Tempo libero'],
    'Spesa': [370, 120, 105,
              340, 100, 90,
              400, 130, 130]
}
df_gastos_mes = pd.DataFrame(data_mes)

# Creiamo la tabella pivot: righe = mese, colonne = categoria, valori = somma della spesa
pivot_gastos = df_gastos_mes.pivot_table(
    index='Mese',
    columns='Categoria',
    values='Spesa',
    aggfunc='sum'
)

# Creiamo il grafico della tabella pivot come barre
pivot_gastos.plot(kind='bar', figsize=(10, 6))
plt.title('Le mie spese per categoria e mese')
plt.xlabel('Mese')
plt.ylabel('Spesa totale')
plt.grid(False)
plt.legend(title='Categoria', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
