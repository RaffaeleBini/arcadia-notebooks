```python
# Datos de ejemplo: gastos personales por mes y categoria
data_mes = {
    'Mes': ['Enero', 'Enero', 'Enero',
            'Febrero', 'Febrero', 'Febrero',
            'Marzo', 'Marzo', 'Marzo'],
    'Categoria': ['Comida', 'Transporte', 'Ocio',
                  'Comida', 'Transporte', 'Ocio',
                  'Comida', 'Transporte', 'Ocio'],
    'Gasto': [370, 120, 105,
              340, 100, 90,
              400, 130, 130]
}
df_gastos_mes = pd.DataFrame(data_mes)

# Creamos la tabla dinamica: filas = mes, columnas = categoria, valores = suma del gasto
pivot_gastos = df_gastos_mes.pivot_table(
    index='Mes',
    columns='Categoria',
    values='Gasto',
    aggfunc='sum'
)

# Graficamos la tabla dinamica como barras
pivot_gastos.plot(kind='bar', figsize=(10, 6))
plt.title('Mis gastos por categoria y mes')
plt.xlabel('Mes')
plt.ylabel('Gasto total')
plt.grid(False)
plt.legend(title='Categoria', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
