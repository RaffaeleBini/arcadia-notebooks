## Esercizio 2 — Mappa del Messico con design moderno/chiaro e zoom livello 6

**Consegna:** Crea una mappa del Messico con un design pulito, moderno e a sfondo chiaro, usando un livello di zoom di 6.

**Suggerimenti:**
- Lo stile di tiles `'Cartodb positron'` è un design minimalista a sfondo chiaro, ideale per questo caso.
- Riusa le stesse coordinate del centro del Messico dell'esercizio precedente.
- Il parametro `tiles` si passa allo stesso modo di `location` e `zoom_start` quando costruisci il `folium.Map(...)`.

````python
# TODO: riusa le coordinate del Messico (mexico_latitude, mexico_longitude)
# TODO: crea una mappa di Folium centrata sul Messico con zoom_start=6 e tiles='Cartodb positron'
# TODO: mostra la mappa
````

<details>

