## 0. Preparazione: librerie e dati

````python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Carichiamo il dataset delle compagnie aeree
airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str}
)

# Prendiamo lo stesso campione di 500 righe usato in lezione
data = airline_data.sample(n=500, random_state=42)

data.shape
````

## Esercizio 1 — Line plot del ritardo medio mensile di arrivo

Crea un **line plot** che mostri, per ogni mese (`Month`), il **ritardo medio di arrivo** (`ArrDelay`).

**Suggerimenti:**
- Raggruppa `data` per `'Month'` e calcola la media di `'ArrDelay'` con `.groupby('Month')['ArrDelay'].mean().reset_index()`.
- Usa `graph_objects`: crea una figura vuota con `go.Figure()`, aggiungi una traccia con `go.Scatter(..., mode='lines')` e chiama `fig.add_trace(...)`.
- Regola titolo ed etichette degli assi con `fig.update_layout(title=..., xaxis_title=..., yaxis_title=...)`.
- Termina con `fig.show()`.

````python
# TODO: raggruppa `data` per 'Month' e calcola la media di 'ArrDelay'
# TODO: crea una figura vuota con go.Figure()
# TODO: aggiungi una traccia go.Scatter con mode='lines' usando Month in x e il ritardo medio in y
# TODO: regola titolo ed etichette degli assi con fig.update_layout()
# TODO: mostra il grafico con fig.show()
````

<details>
<summary>💡 Vedi soluzione</summary>

```python
# Raggruppiamo per mese e calcoliamo il ritardo medio di arrivo
line_data = data.groupby('Month')['ArrDelay'].mean().reset_index()

# Passo 1: creiamo una figura vuota
fig = go.Figure()

# Passo 2: aggiungiamo una traccia di tipo linea (mode='lines' invece di 'markers')
fig.add_trace(go.Scatter(
    x=line_data['Month'],
    y=line_data['ArrDelay'],
    mode='lines',
    marker=dict(color='green')
))

# Passo 3: titolo ed etichette degli assi
fig.update_layout(
    title='Month vs Average Flight Delay Time',
    xaxis_title='Month',
    yaxis_title='ArrDelay'
)

fig.show()
```

````python
# TODO: raggruppa `data` per 'DestState' e somma 'Flights'
# TODO: crea il bar chart con px.bar(), passando x="DestState", y="Flights" e un titolo
# TODO: mostra il grafico con fig.show()
````

# Raggruppiamo per stato di destinazione e sommiamo il totale dei voli
bar_data = data.groupby(['DestState'])['Flights'].sum().reset_index()

# Una sola riga: passiamo il dataframe, x, y e il titolo
