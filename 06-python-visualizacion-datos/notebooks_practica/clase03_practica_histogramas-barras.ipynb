```python
df_india = df_can.loc['India', years]

df_india.plot(kind='bar', figsize=(10, 6), rot=90)

plt.xlabel('Año')
plt.ylabel('Número de inmigrantes')
plt.title('Inmigrantes indios a Canadá (1980-2013)')

# posición (índice numérico) y valor del pico máximo
peak_index = df_india.values.argmax()
peak_value = df_india.values.max()

# la flecha apunta desde un poco antes del pico hacia el pico mismo
plt.annotate('',
             xy=(peak_index, peak_value),
             xytext=(peak_index - 4, peak_value - 5000),
             xycoords='data',
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='blue', lw=2))

plt.annotate(f'Máximo: {peak_value} en {df_india.index[peak_index]}',
             xy=(peak_index - 4, peak_value - 4500),
             va='bottom',
             ha='left')

plt.show()
```

````python
# TODO: ordena df_can por 'Total' ascendente y toma los 5 países con menos inmigración total
# TODO: grafica con kind='barh'
# TODO: añade xlabel y title
# TODO: recorre la serie con enumerate() y usa plt.annotate para poner el valor junto a cada barra
# TODO: plt.show()
````

<details>

```python
# paso 1: ordenar ascendente y quedarnos con los 5 menores
df_can.sort_values(by='Total', ascending=True, inplace=True)
df_bottom5 = df_can['Total'].head(5)

# paso 2: graficar barras horizontales
df_bottom5.plot(kind='barh', figsize=(10, 6), color='steelblue')
plt.xlabel('Número de inmigrantes')
plt.title('Los 5 países que menos contribuyeron a la inmigración a Canadá (1980-2013)')

# paso 3: anotar el valor exacto junto a cada barra
for index, value in enumerate(df_bottom5):
    label = format(int(value), ',')
    plt.annotate(label, xy=(value + 0.5, index), va='center', color='black')

plt.show()
```
</details>

## Recapitulación

- Un histograma multi-serie necesita datos **transpuestos**: países/series como columnas, años como filas.
- `argmax()` sobre `.values` te da la posición del máximo dentro de una serie de pandas.
- `enumerate()` es la forma idiomática de anotar cada barra de un `barh` con su valor exacto.
