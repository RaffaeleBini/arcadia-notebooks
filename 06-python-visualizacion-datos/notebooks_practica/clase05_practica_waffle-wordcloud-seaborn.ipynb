```python
years = list(map(str, range(1980, 2014)))

# crear df_countries: países en columnas, años en filas
df_countries = df_can.loc[['Denmark', 'Norway', 'Sweden'], years].transpose()

# sumar a través de los tres países para cada año
df_total = pd.DataFrame(df_countries.sum(axis=1))

# resetear índice para tener 'year' como columna
df_total.reset_index(inplace=True)

# renombrar columnas
df_total.columns = ['year', 'total']

# convertir year de string a int para el gráfico
df_total['year'] = df_total['year'].astype(int)

# definir tamaño de figura
plt.figure(figsize=(15, 10))

# definir estilo y tamaño de fuente
sns.set(font_scale=1.5)
sns.set_style('whitegrid')

# generar el regplot con título y ejes etiquetados
ax = sns.regplot(
    x='year',
    y='total',
    data=df_total,
    color='green',
    marker='+',
    scatter_kws={'s': 200}
)

ax.set(xlabel='Year', ylabel='Total Immigration')
ax.set_title('Total Immigration from Denmark, Sweden, and Norway to Canada from 1980 - 2013')
plt.show()
```
