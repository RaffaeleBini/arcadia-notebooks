# Ordenamos por 'Total' ascendente para obtener los que menos aportaron y tomamos los primeros 5
df_least5 = df_can.sort_values(by='Total', ascending=True, axis=0).head(5)

# Nos quedamos solo con las columnas de años y transponemos para graficar bien
df_least5 = df_least5[years].transpose()

df_least5.index = df_least5.index.map(int)

# Versión apilada (comportamiento por defecto de area plot)
df_least5.plot(kind='area', stacked=True, figsize=(14, 8))
plt.title('Tendencia de los 5 Países con Menos Inmigración (apilado)')
plt.ylabel('Número de Inmigrantes')
plt.xlabel('Años')
plt.show()

# Versión sin apilar, con transparencia para ver las áreas superpuestas
df_least5.plot(kind='area', stacked=False, alpha=0.5, figsize=(14, 8))
plt.title('Tendencia de los 5 Países con Menos Inmigración (sin apilar)')
plt.ylabel('Número de Inmigrantes')
plt.xlabel('Años')
plt.show()

# La versión apilada suma las áreas: el borde superior representa el total combinado,
# no la magnitud individual de cada país. La versión sin apilar (stacked=False) permite
# comparar la tendencia real de cada país por separado, aunque las áreas se superpongan.
