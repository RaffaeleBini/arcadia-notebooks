## Ejercicio 1 — Pie chart: proporción de inmigrantes por continente en 2013

Crea un pie chart que muestre la **proporción de inmigrantes por continente** correspondiente solo al año **2013** (no el total 1980-2013). Usa `explode` para resaltar los **tres continentes con menor aporte**.

**Pistas:**
- Agrupa `df_can` por `'Continent'` con `.groupby('Continent').sum(numeric_only=True)` para obtener `df_continents`.
- La columna de interés ya no es `'Total'`, sino la columna del año `'2013'`.
- Ordena los continentes por el valor de 2013 (`.sort_values()`) para saber cuáles son los tres con menor aporte y así construir la lista `explode_list` (un valor por continente, en el mismo orden que el índice del DataFrame que grafiques).
- Usa `autopct='%1.1f%%'`, `startangle=90` y `plt.axis('equal')` como en la teoría.

````python
# TODO: agrupa df_can por 'Continent' sumando valores numéricos -> df_continents

# TODO: identifica los tres continentes con menor valor en la columna '2013'

# TODO: construye explode_list (un valor por continente: p.ej. 0.1 para los tres menores, 0 para el resto)
# respetando el ORDEN de df_continents.index

# TODO: dibuja el pie chart usando df_continents['2013'] con autopct, startangle, explode y leyenda

# TODO: plt.axis('equal') y plt.show()
````

<details>
<summary>💡 Ver solución</summary>

```python
# agrupar por continente, sumando todas las columnas numéricas (incluye la columna '2013')
df_continents = df_can.groupby('Continent').sum(numeric_only=True)

# ordenar continentes por el valor de 2013, de menor a mayor
orden_2013 = df_continents['2013'].sort_values()

# los tres continentes con menor aporte en 2013
menores = set(orden_2013.index[:3])

# construir explode_list respetando el orden real de df_continents.index
explode_list = [0.1 if continente in menores else 0 for continente in df_continents.index]

df_continents['2013'].plot(kind='pie',
                            figsize=(10, 6),
                            autopct='%1.1f%%',
                            startangle=90,
                            shadow=True,
                            pctdistance=1.12,
                            explode=explode_list)

plt.title('Inmigración a Canadá por continente (2013)', y=1.12, fontsize=15)
plt.axis('equal')
plt.legend(labels=df_continents.index, loc='upper left', fontsize=8)
plt.show()
```

## Ejercicio 2 — Bubble plot: China vs. India (normalizado)

Construye un bubble plot que compare la inmigración de **China** e **India** (1980-2013), usando el tamaño de la burbuja para representar la magnitud de cada año. Cada país debe **normalizarse por separado** (feature scaling al rango [0, 1]) antes de graficar, igual que se hizo en la teoría con Brasil y Argentina.

**Pistas:**
- Parte de `df_can_t = df_can[years].transpose()`, convierte el índice a entero y resetéalo como columna `'Year'` (igual que en la teoría).
- Normaliza cada serie con la fórmula `(x - x.min()) / (x.max() - x.min())`.
- Escala el resultado normalizado para el parámetro `s`, por ejemplo `norm * 2000 + 10`, para que ningún punto sea invisible.
- Dibuja primero China con `df_can_t.plot(kind='scatter', ...)`, guarda el eje devuelto (`ax0`), y vuelve a llamar `.plot()` para India pasando `ax=ax0` para que ambos países compartan los mismos ejes.

````python
# TODO: construye df_can_t = df_can[years].transpose(), con índice entero llamado 'Year' reseteado como columna

# TODO: normaliza 'China' -> norm_china, y 'India' -> norm_india, al rango [0, 1]

# TODO: dibuja el scatter de China con s=norm_china * 2000 + 10, guarda el eje en ax0

# TODO: dibuja el scatter de India con s=norm_india * 2000 + 10, ax=ax0

# TODO: añade título, etiqueta del eje y y leyenda ['China', 'India']; plt.show()
````

```python
# transponer para tener los años como filas
df_can_t = df_can[years].transpose()
df_can_t.index = map(int, df_can_t.index)
df_can_t.index.name = 'Year'
df_can_t.reset_index(inplace=True)

# normalizar China e India por separado (feature scaling a [0, 1])
norm_china = (df_can_t['China'] - df_can_t['China'].min()) / (df_can_t['China'].max() - df_can_t['China'].min())
norm_india = (df_can_t['India'] - df_can_t['India'].min()) / (df_can_t['India'].max() - df_can_t['India'].min())

# China
ax0 = df_can_t.plot(kind='scatter',
                     x='Year',
                     y='China',
                     figsize=(14, 8),
                     alpha=0.5,
                     color='green',
                     s=norm_china * 2000 + 10,
                     xlim=(1975, 2015))

# India, en los mismos ejes que China
ax1 = df_can_t.plot(kind='scatter',
                     x='Year',
                     y='India',
                     alpha=0.5,
                     color='blue',
                     s=norm_india * 2000 + 10,
                     ax=ax0)

ax0.set_ylabel('Número de inmigrantes')
ax0.set_title('Inmigración desde China e India a Canadá (1980-2013)')
ax0.legend(['China', 'India'], loc='upper left', fontsize='x-large')

plt.show()
```
