```python
# Ordiniamo per 'Total' crescente per ottenere quelli che hanno contribuito meno e prendiamo i primi 5
df_least5 = df_can.sort_values(by='Total', ascending=True, axis=0).head(5)

# Teniamo solo le colonne degli anni e trasponiamo per creare il grafico correttamente
df_least5 = df_least5[years].transpose()

# Convertiamo l'indice in intero per un asse X pulito
df_least5.index = df_least5.index.map(int)

# Versione impilata (comportamento di default dell'area plot)
df_least5.plot(kind='area', stacked=True, figsize=(14, 8))
plt.title('Tendenza dei 5 paesi con meno immigrazione (impilato)')
plt.ylabel('Numero di immigrati')
plt.xlabel('Anni')
plt.show()

# Versione non impilata, con trasparenza per vedere le aree sovrapposte
df_least5.plot(kind='area', stacked=False, alpha=0.5, figsize=(14, 8))
plt.title('Tendenza dei 5 paesi con meno immigrazione (senza impilare)')
plt.ylabel('Numero di immigrati')
plt.xlabel('Anni')
plt.show()

# La versione impilata somma le aree: il bordo superiore rappresenta il totale combinato,
# non la grandezza individuale di ogni paese. La versione non impilata (stacked=False) permette
# di confrontare la tendenza reale di ogni paese separatamente, anche se le aree si sovrappongono.
```
