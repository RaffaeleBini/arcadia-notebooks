## Esercizio 1 — Cambia il titolo e la dimensione del font

**Consegna:** Parti dal codice base della dashboard delle compagnie aeree (blocco 2, `app.layout`). Fai due modifiche:

1. Cambia il testo del titolo da `'Airline Dashboard'` a `'Airline On-time Performance Dashboard'`.
2. Cambia il `font-size` del titolo da `40` a `50`.

**Suggerimenti:**

- Il titolo si definisce dentro `html.H1(...)`. Il primo argomento posizionale è il testo; lo `style` è un dizionario dove vive `font-size`.
- Non devi toccare nient'altro del layout — né il paragrafo (`html.P`) né il grafico (`dcc.Graph`) cambiano.
- Se lo esegui come `.py`, ricorda di salvare il file e rieseguire `python file.py` per vedere la modifica; se usi `jupyter_mode='inline'`, basta rieseguire la cella.

````python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str})

data = airline_data.sample(n=500, random_state=42)

fig = px.pie(data, values='Flights', names='DistanceGroup',
             title='Distance group proportion by flights')

app = dash.Dash(__name__)

# TODO: cambia il testo del titolo in 'Airline On-time Performance Dashboard'
# TODO: cambia il 'font-size' del titolo da 40 a 50
app.layout = html.Div(children=[
    html.H1('Airline Dashboard',
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    html.P('Proportion of distance group (250 mile distance interval group) by flights.',
           style={'textAlign': 'center', 'color': '#F57241'}),
    dcc.Graph(figure=fig),
])

if __name__ == '__main__':
    app.run()
````

```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str})

data = airline_data.sample(n=500, random_state=42)

# Pie chart originale
fig = px.pie(data, values='Flights', names='DistanceGroup',
             title='Distance group proportion by flights')

# Secondo grafico: barre di voli per gruppo di distanza
fig2 = px.bar(data, x='DistanceGroup', y='Flights',
              title='Flights by distance group')

app = dash.Dash(__name__)

app.layout = html.Div(children=[
    html.H1('Airline Dashboard',
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    html.P('Proportion of distance group (250 mile distance interval group) by flights.',
           style={'textAlign': 'center', 'color': '#F57241'}),
    dcc.Graph(figure=fig),
    # Secondo grafico aggiunto sotto il primo, nella stessa lista children
    dcc.Graph(figure=fig2),
])

if __name__ == '__main__':
    app.run()
```
