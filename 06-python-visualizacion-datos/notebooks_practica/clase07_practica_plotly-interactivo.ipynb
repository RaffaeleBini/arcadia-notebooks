## 0. Preparación: librerías y datos

````python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Cargamos el dataset de aerolíneas
airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str}
)

# Tomamos la misma muestra de 500 filas que usamos en la clase
data = airline_data.sample(n=500, random_state=42)

data.shape
````

- Ajusta el título y las etiquetas de los ejes con `fig.update_layout(title=..., xaxis_title=..., yaxis_title=...)`.
- Termina con `fig.show()`.

````python
# TODO: agrupa `data` por 'Month' y calcula la media de 'ArrDelay'
# TODO: crea una figura vacía con go.Figure()
# TODO: añade una traza go.Scatter con mode='lines' usando Month en x y el retraso promedio en y
# TODO: ajusta título y etiquetas de los ejes con fig.update_layout()
# TODO: muestra el gráfico con fig.show()
````

<details>
<summary>💡 Ver solución</summary>

```python
# Agrupamos por mes y calculamos el retraso promedio de llegada
line_data = data.groupby('Month')['ArrDelay'].mean().reset_index()

# Paso 1: creamos una figura vacía
fig = go.Figure()

# Paso 2: añadimos una traza de tipo línea (mode='lines' en vez de 'markers')
fig.add_trace(go.Scatter(
    x=line_data['Month'],
    y=line_data['ArrDelay'],
    mode='lines',
    marker=dict(color='green')
))

# Paso 3: título y etiquetas de los ejes
fig.update_layout(
    title='Month vs Average Flight Delay Time',
    xaxis_title='Month',
    yaxis_title='ArrDelay'
)

fig.show()
```

## Ejercicio 2 — Bar chart del número de vuelos por estado destino

Crea un **bar chart** con `plotly.express` que muestre el número total de vuelos (`Flights`) por estado de destino (`DestState`).

**Pistas:**
- Agrupa `data` por `'DestState'` y suma `'Flights'` con `.groupby(['DestState'])['Flights'].sum().reset_index()`.
- Usa `px.bar(dataframe, x="DestState", y="Flights", title=...)`.
- Termina con `fig.show()`.

````python
# TODO: agrupa `data` por 'DestState' y suma 'Flights'
# TODO: crea el bar chart con px.bar(), pasando x="DestState", y="Flights" y un título
# TODO: muestra el gráfico con fig.show()
````

```python
# Agrupamos por estado de destino y sumamos el total de vuelos
bar_data = data.groupby(['DestState'])['Flights'].sum().reset_index()

# Una sola línea: pasamos el dataframe, x, y y el título
fig = px.bar(bar_data, x="DestState", y="Flights",
             title='Total number of flights to the destination state split by reporting airline')
```
