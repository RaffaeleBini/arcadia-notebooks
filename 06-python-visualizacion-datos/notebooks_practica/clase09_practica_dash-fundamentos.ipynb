## Ejercicio 1 — Cambia el título y el tamaño de fuente

**Enunciado:** Parte del código base del dashboard de aerolíneas (bloque 2, `app.layout`). Haz dos cambios:

1. Cambia el texto del título de `'Airline Dashboard'` a `'Airline On-time Performance Dashboard'`.
2. Cambia el `font-size` del título de `40` a `50`.

**Pistas:**

- El título se define dentro de `html.H1(...)`. El primer argumento posicional es el texto; el `style` es un diccionario donde vive `font-size`.
- No necesitas tocar nada más del layout — ni el párrafo (`html.P`) ni el gráfico (`dcc.Graph`) cambian.
- Si ejecutas esto como `.py`, recuerda guardar el archivo y volver a correr `python archivo.py` para ver el cambio; si usas `jupyter_mode='inline'`, basta con volver a ejecutar la celda.

````python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str})

data = airline_data.sample(n=500, random_state=42)

fig = px.pie(data, values='Flights', names='DistanceGroup',
             title='Distance group proportion by flights')

app = dash.Dash(__name__)

# TODO: cambia el texto del título a 'Airline On-time Performance Dashboard'
# TODO: cambia el 'font-size' del título de 40 a 50
app.layout = html.Div(children=[
    html.H1('Airline Dashboard',
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    html.P('Proportion of distance group (250 mile distance interval group) by flights.',
           style={'textAlign': 'center', 'color': '#F57241'}),
    dcc.Graph(figure=fig),
])

if __name__ == '__main__':
    app.run()
````

```python
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

airline_data = pd.read_csv(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv',
    encoding="ISO-8859-1",
    dtype={'Div1Airport': str, 'Div1TailNum': str,
           'Div2Airport': str, 'Div2TailNum': str})

data = airline_data.sample(n=500, random_state=42)

# Pie chart original
fig = px.pie(data, values='Flights', names='DistanceGroup',
             title='Distance group proportion by flights')

# Segundo gráfico: barras de vuelos por grupo de distancia
fig2 = px.bar(data, x='DistanceGroup', y='Flights',
              title='Flights by distance group')

app = dash.Dash(__name__)

app.layout = html.Div(children=[
    html.H1('Airline Dashboard',
            style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
    html.P('Proportion of distance group (250 mile distance interval group) by flights.',
           style={'textAlign': 'center', 'color': '#F57241'}),
    dcc.Graph(figure=fig),
    # Segundo gráfico añadido debajo del primero, en la misma lista children
    dcc.Graph(figure=fig2),
])

if __name__ == '__main__':
    app.run()
```
