**Contexto de negocio:** la dirección de XYZAutomotives sospecha que, cuando la confianza del consumidor cae durante una recesión, las ventas caen todavía más — y quiere saber si esa relación es real antes de decidir si vale la pena invertir en campañas que intenten "levantar el ánimo" del comprador en plena crisis. Te piden un gráfico que muestre, **solo para los periodos de recesión** (`Recession == 1`), la relación entre `Consumer_Confidence` y `Automobile_Sales`.

**Pistas:**
- Filtra el dataframe para quedarte solo con las filas en recesión (`Recession == 1`).
- Un scatter plot es el gráfico correcto aquí: estás relacionando dos variables numéricas continuas, no comparando categorías ni viendo una tendencia temporal.
- Puedes usar `plt.scatter()` o `sns.scatterplot()`.
- No olvides `xlabel`, `ylabel` y `title`.

````python
# Cargamos el dataset de ventas de automóviles
import requests
import io

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/d51iMGfp_t0QpO30Lym-dw/automobile-sales.csv"
response = requests.get(URL)
response.raise_for_status()
df_autos = pd.read_csv(io.StringIO(response.text))

df_autos.head()
````

````python
# TODO: filtra df_autos para quedarte solo con las filas donde Recession == 1
# TODO: crea un scatter plot de Consumer_Confidence (eje X) vs Automobile_Sales (eje Y)
# TODO: añade xlabel, ylabel y un title descriptivo, por ejemplo:
#       'Confianza del consumidor vs. ventas de automóviles durante recesión'
````

## Ejercicio 2 — Brillo medio de incendio por región (Australia)

**Contexto de negocio:** el organismo de monitoreo ambiental quiere saber si alguna de las siete regiones de Australia tiende a registrar incendios de mayor intensidad (no solo más frecuentes) que las demás, medido por `Mean_estimated_fire_brightness`. Esa información ayuda a decidir dónde priorizar equipos de detección temprana de alta sensibilidad, distinto de simplemente mandar más recursos a donde hay más incendios en número.

**Pistas:**
- Aquí estás comparando una variable numérica (`Mean_estimated_fire_brightness`) entre categorías con nombre (las siete regiones) — el trabajo natural de un gráfico de barras, no de una línea ni de un histograma.
- Usa `sns.barplot(data=..., x='Region', y='Mean_estimated_fire_brightness')`. Por defecto, `seaborn` calcula automáticamente la media y muestra un intervalo de confianza — no necesitas agrupar tú mismo con `groupby`.
- No olvides `xlabel`, `ylabel` y `title`.

````python
# Cargamos el dataset de incendios forestales en Australia
URL_FIRES = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Historical_Wildfires.csv"
response = requests.get(URL_FIRES)
response.raise_for_status()
df_fires = pd.read_csv(io.StringIO(response.text))

df_fires.head()
````

````python
# TODO: usa sns.barplot() para comparar Mean_estimated_fire_brightness entre las regiones (columna 'Region')
# TODO: añade xlabel, ylabel y un title descriptivo, por ejemplo:
#       'Brillo medio de incendio por región'
````

<details>
<summary>💡 Ver solución</summary>

```python
# Comparación categórica (7 regiones) de una variable numérica: el trabajo natural de las barras
plt.figure(figsize=(10, 6))
sns.barplot(data=df_fires, x='Region', y='Mean_estimated_fire_brightness')

plt.xlabel('Región')
plt.ylabel('Brillo medio de incendio (Kelvin)')
plt.title('Brillo medio de incendio por región')
plt.tight_layout()
plt.show()

# Lectura: la región con la barra más alta no es necesariamente la que más incendios tiene
# (eso lo verías con un pie de Count por región), sino la que registra los incendios más
# intensos en promedio. Esa distinción es justo la que le importa al organismo de monitoreo
# para decidir dónde reforzar la detección temprana de alta sensibilidad.
```
