## Esercizio 1 — Fiducia dei consumatori e vendite durante la recessione (automobili)

**Contesto di business:** la direzione di XYZAutomotives sospetta che, quando la fiducia dei consumatori cala durante una recessione, le vendite calino ancora di più — e vuole sapere se questa relazione è reale prima di decidere se vale la pena investire in campagne che cerchino di "sollevare il morale" dell'acquirente in piena crisi. Ti chiedono un grafico che mostri, **solo per i periodi di recessione** (`Recession == 1`), la relazione tra `Consumer_Confidence` e `Automobile_Sales`.

**Suggerimenti:**
- Filtra il dataframe per tenere solo le righe in recessione (`Recession == 1`).
- Uno scatter plot è il grafico giusto qui: stai collegando due variabili numeriche continue, non stai confrontando categorie né guardando una tendenza temporale.
- Puoi usare `plt.scatter()` o `sns.scatterplot()`.
- Non dimenticare `xlabel`, `ylabel` e `title`.

````python
# Carichiamo il dataset delle vendite di automobili
import requests
import io

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/d51iMGfp_t0QpO30Lym-dw/automobile-sales.csv"
response = requests.get(URL)
response.raise_for_status()
df_autos = pd.read_csv(io.StringIO(response.text))

df_autos.head()
````

````python
# TODO: filtra df_autos per tenere solo le righe dove Recession == 1
# TODO: crea uno scatter plot di Consumer_Confidence (asse X) vs Automobile_Sales (asse Y)
# TODO: aggiungi xlabel, ylabel e un title descrittivo, ad esempio:
#       'Fiducia del consumatore vs. vendite di automobili durante la recessione'
````

```python
# Filtriamo solo i periodi di recessione
rec_data = df_autos[df_autos['Recession'] == 1]

# Scatter plot: due variabili numeriche continue, senza dimensione temporale né categorie
plt.figure(figsize=(8, 6))
plt.scatter(rec_data['Consumer_Confidence'], rec_data['Automobile_Sales'], alpha=0.6, color='darkred')

plt.title('Fiducia del consumatore vs. vendite di automobili durante la recessione')

plt.show()

# Lettura: se i punti salgono da sinistra a destra (correlazione positiva),
# questo conferma il sospetto della direzione: a minore fiducia del consumatore,
# minori vendite anche all'interno dello stesso periodo di recessione. Questo giustificherebbe
# investire in campagne che rafforzino la fiducia dell'acquirente durante la crisi.
```

## Esercizio 2 — Luminosità media dell'incendio per regione (Australia)

**Contesto di business:** l'ente di monitoraggio ambientale vuole sapere se qualcuna delle sette regioni dell'Australia tende a registrare incendi di maggiore intensità (non solo più frequenti) rispetto alle altre, misurato tramite `Mean_estimated_fire_brightness`. Questa informazione aiuta a decidere dove dare priorità a squadre di rilevazione precoce ad alta sensibilità, diverso dal semplicemente mandare più risorse dove ci sono più incendi in numero.

**Suggerimenti:**
- Qui stai confrontando una variabile numerica (`Mean_estimated_fire_brightness`) tra categorie con nome (le sette regioni) — il lavoro naturale di un grafico a barre, non di una linea né di un istogramma.
- Usa `sns.barplot(data=..., x='Region', y='Mean_estimated_fire_brightness')`. Di default, `seaborn` calcola automaticamente la media e mostra un intervallo di confidenza — non devi raggruppare tu stesso con `groupby`.
- Non dimenticare `xlabel`, `ylabel` e `title`.

````python
# Carichiamo il dataset degli incendi boschivi in Australia
URL_FIRES = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Historical_Wildfires.csv"
response = requests.get(URL_FIRES)
response.raise_for_status()
df_fires = pd.read_csv(io.StringIO(response.text))

df_fires.head()
````

````python
# TODO: usa sns.barplot() per confrontare Mean_estimated_fire_brightness tra le regioni (colonna 'Region')
# TODO: aggiungi xlabel, ylabel e un title descrittivo, ad esempio:
#       'Luminosità media dell'incendio per regione'
````

<details>
<summary>💡 Vedi soluzione</summary>

```python
# Confronto categorico (7 regioni) di una variabile numerica: il lavoro naturale delle barre
plt.figure(figsize=(10, 6))
sns.barplot(data=df_fires, x='Region', y='Mean_estimated_fire_brightness')

plt.xlabel('Regione')
plt.ylabel('Luminosità media dell\'incendio (Kelvin)')
plt.title('Luminosità media dell\'incendio per regione')
plt.tight_layout()
plt.show()

# Lettura: la regione con la barra più alta non è necessariamente quella con più incendi
# (lo vedresti con un pie di Count per regione), ma quella che registra gli incendi più
# intensi in media. Questa distinzione è proprio quella che interessa all'ente di monitoraggio
# per decidere dove rafforzare la rilevazione precoce ad alta sensibilità.
```
