## Caso 1 — XYZAutomotives: ventas de coches y recesión económica

**La pregunta de negocio:** ¿cómo golpean las recesiones económicas a nuestras ventas? La dirección necesita saberlo para planificar producción, inventario y publicidad cuando ve venir una crisis.

El dataset cubre 1980-2013(+) y marca seis periodos de recesión reales: 1980, 1981-1982, 1991, 2000-2001, finales de 2007 a mediados de 2009 (crisis financiera) y 2020 (Covid-19). Además de ventas mensuales, trae PIB, desempleo, confianza del consumidor, gasto en publicidad, precio medio del vehículo y tipo de coche vendido.

**Qué gráficos responden qué:**

- **Línea** para ventas por año: hay una dimensión temporal continua de más de 30 años — la línea es la única forma de ver tendencia y quiebres de un vistazo. Barras aquí serían ilegibles.
- **Barras** para comparar ventas medias por tipo de vehículo en recesión vs. no recesión: son categorías con nombre, sin orden temporal — el caso de libro para barras.
- **Pie** para el reparto del gasto publicitario por tipo de vehículo: la pregunta es "¿qué proporción del total se lleva cada categoría?", y con solo cinco categorías el pie sigue siendo legible.
- **Barras apiladas por color** para el efecto del desempleo sobre las ventas por tipo de coche: combina dos dimensiones categóricas sin recurrir a un gráfico 3D.
- **Scatter/bubble** para relacionar confianza del consumidor o precio con volumen de ventas: dos variables numéricas continuas, el trabajo del punto de dispersión.

**La decisión que permite tomar:** un director puede ver si las ventas de coches deportivos cayeron más que las familiares durante una recesión concreta, y decidir si reforzar publicidad en el segmento más resiliente la próxima vez que las señales de PIB y desempleo anticipen una crisis.

## Caso 2 — Aerolíneas: retrasos de vuelo por causa y por mes

**La pregunta de negocio:** ¿por qué se retrasan los vuelos, no solo cuánto? Un retraso por mala gestión de tripulación (`carrier delay`) exige una respuesta distinta a uno por tormenta (`weather delay`), por congestión del sistema aéreo (`NAS delay`), por seguridad (`security delay`) o porque el avión llegó tarde del vuelo anterior (`late aircraft delay`).

**Qué gráfico se usó y por qué:** cinco paneles, todos de **líneas**, una línea de color por aerolínea (`color='Reporting_Airline'`). La dimensión temporal (el mes) y la necesidad de ver tendencia estacional descartan las barras — con doce meses y una docena de aerolíneas, un gráfico de barras agrupadas sería un muro de color ilegible.

```python
carrier_fig = px.line(avg_car, x='Month', y='CarrierDelay',
                       color='Reporting_Airline',
                       title='Average carrier delay time (minutes) by airline')
```

**La decisión que permite tomar:** un gestor de operaciones puede ver si el retraso por "late aircraft" de una aerolínea se dispara en meses de más tráfico, señal de un problema de rotación de flota — mientras que si el problema fuera `weather delay`, la solución sería logística de contingencia, no de programación.

## Caso 3 — Incendios forestales en Australia

**La pregunta de negocio:** ¿qué regiones concentran más actividad de incendios? ¿hay estacionalidad? ¿ha ido a más o a menos con los años? Un organismo de monitoreo ambiental recoge, desde 2005, datos diarios por cada una de las siete regiones de Australia: área estimada de incendio, brillo medio del fuego, potencia radiativa, confianza de la detección y conteo de píxeles satelitales de vegetación en llamas.

**Qué gráficos se usaron y por qué:**

- **Línea** para el área de incendio a lo largo del tiempo — dimensión temporal.
- **Barras (seaborn)** para comparar el brillo medio del fuego entre las siete regiones — categorías sin orden temporal.
- **Pie** para la proporción del conteo de píxeles por región — pocas categorías, reparto de un total.
- **Histograma** para la distribución del brillo medio del fuego — aquí la pregunta ya no es "cómo se compara una región con otra" sino "qué tan común es cada nivel de brillo", terreno de histograma.
- **Scatter** para relacionar potencia radiativa media con nivel de confianza medio — dos variables numéricas continuas.
- **Mapa con folium** marcando las siete regiones — la única de las tres historias con componente geográfico central a la pregunta ("¿dónde ocurre esto?").

**La decisión que permite tomar:** un analista ambiental puede ver qué región concentra la mayor proporción de píxeles en llamas en un mes dado y priorizar el envío de recursos de extinción hacia esa región, en vez de repartirlos por igual entre las siete.

## El hilo común

Fíjate en lo que decide cada gráfico:

| Pregunta | Tipo de gráfico | Por qué |
|---|---|---|
| ¿Sube o baja con el tiempo? | Línea | Dimensión temporal continua |
| ¿Cómo se comparan categorías con nombre? | Barras | Categorías discretas, sin orden temporal |
| ¿Qué proporción de un total tiene cada categoría? | Pie | Pocas categorías, reparto de un total |
| ¿Qué tan común es cada valor? | Histograma | Distribución de una variable numérica |
| ¿Cómo se relacionan dos variables numéricas? | Scatter / bubble | Relación entre dos (o tres) variables continuas |
| ¿Dónde ocurre esto? | Mapa (folium) | Componente geográfico central a la pregunta |

Un dashboard real casi nunca usa un solo tipo de gráfico: combina varios de la tabla de arriba según la pregunta que responde cada panel. Antes de elegir un tipo de gráfico, pregúntate siempre **quién lo va a mirar y qué decisión tiene que tomar después de verlo** — esa pregunta, no la variedad de gráficos que conoces, es la que guía el diseño de un dashboard útil.

## Preparar el entorno y cargar los datos

Usamos el mismo dataset `automobile-sales.csv` del proyecto integrador de automóviles y recesión: ventas mensuales entre 1980 y 2013+ con variables económicas (PIB, desempleo, confianza del consumidor), publicidad, precio y tipo de vehículo.

````python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

print('Librerías importadas correctamente')
````

````python
import requests
import io

# URL del dataset de ventas de automóviles
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/d51iMGfp_t0QpO30Lym-dw/automobile-sales.csv"

response = requests.get(URL)
response.raise_for_status()

csv_content = io.StringIO(response.text)
df = pd.read_csv(csv_content)

print('Datos descargados y leídos en un dataframe!')
df.head()
````

## Paso 1 — Línea de ventas por año, con anotaciones de recesión

Empezamos por la pregunta más básica: ¿cómo evolucionan las ventas medias de automóviles a lo largo de los años? Agrupamos por `Year` y calculamos la media de `Automobile_Sales`, luego anotamos sobre el propio gráfico los seis periodos de recesión históricos para que quien lo mire pueda relacionar directamente las caídas de ventas con la economía real.

````python
df_line = df.groupby('Year')['Automobile_Sales'].mean()

plt.figure(figsize=(12, 6))
df_line.plot(kind='line')

plt.xticks(list(range(1980, int(df['Year'].max()) + 1)), rotation=75)
plt.xlabel('Año')
plt.ylabel('Ventas medias de automóviles')
plt.title('Ventas de automóviles durante periodos de recesión')

# Anotamos los seis periodos de recesión históricos sobre el propio gráfico
plt.text(1980, df_line.max() * 0.05, '1980', rotation=90, fontsize=8, color='red')
plt.text(1981, df_line.max() * 0.05, '1981-82', rotation=90, fontsize=8, color='red')
plt.text(1991, df_line.max() * 0.05, '1991', rotation=90, fontsize=8, color='red')
plt.text(2000, df_line.max() * 0.05, '2000-01', rotation=90, fontsize=8, color='red')
plt.text(2007, df_line.max() * 0.05, '2007-09', rotation=90, fontsize=8, color='red')
plt.text(2020, df_line.max() * 0.05, '2020', rotation=90, fontsize=8, color='red')

plt.legend(['Ventas medias'])
plt.tight_layout()
plt.show()
````

**Lectura del gráfico:** cada línea vertical roja marca el inicio de una recesión real. De un vistazo se ve si las ventas caen justo en esos años y con qué fuerza se recuperan después — esa es la pregunta de negocio que motivó todo el proyecto.

## Paso 2 — Barras: ventas medias por tipo de vehículo, recesión vs. no recesión

Ahora bajamos un nivel: no solo "¿caen las ventas en recesión?" sino "¿qué tipo de coche aguanta mejor la caída?". Aquí no hay dimensión temporal continua — hay cinco categorías con nombre (`Vehicle_Type`) cruzadas con una variable binaria (`Recession`) — el caso de libro para un gráfico de barras agrupadas.

````python
grouped_df = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Recession', y='Automobile_Sales', hue='Vehicle_Type', data=grouped_df)

plt.xticks(ticks=[0, 1], labels=['No recesión', 'Recesión'])
plt.xlabel('Periodo')
plt.ylabel('Ventas medias de automóviles')
plt.title('Ventas por tipo de vehículo: recesión vs. no recesión')
plt.legend(title='Tipo de vehículo', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
````

**Lectura del gráfico:** en general todas las categorías caen durante la recesión, pero no todas caen igual. Si `SuperMiniCar` cae menos que `Sports` o `ExecutiveCar`, eso es evidencia directa de que el segmento económico es más resiliente al ciclo económico — la misma conclusión a la que llega el manual del capítulo.

## Paso 3 — Pie: gasto publicitario por tipo de vehículo

Por último, ¿en qué tipo de coche concentra la empresa su publicidad? Con solo cinco categorías y la pregunta siendo "qué proporción del total se lleva cada una", el gráfico de pie sigue siendo legible — con quince categorías ya no lo sería, y ahí tocaría usar barras.

````python
vt_expenditure = df.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()

plt.figure(figsize=(8, 6))
plt.pie(vt_expenditure.values, labels=vt_expenditure.index, autopct='%1.1f%%', startangle=90)
plt.title('Gasto publicitario por tipo de vehículo')
plt.axis('equal')
plt.show()
````

**Lectura del gráfico:** si el segmento económico (`SmallFamilyCar` o `SuperMiniCar`) concentra la mayor porción del gasto publicitario, eso sugiere que la empresa ya apuesta por el segmento más resiliente — coherente con lo que vimos en el gráfico de barras del paso anterior. Los tres gráficos juntos —línea, barras, pie— cuentan una sola historia completa: cuándo cae la venta, qué categoría resiste mejor, y dónde pone la empresa su dinero para reforzarla.
