## Caso 1 — XYZAutomotives: vendite di automobili e recessione economica

**La domanda di business:** come colpiscono le recessioni economiche le nostre vendite? La direzione ha bisogno di saperlo per pianificare produzione, inventario e pubblicità quando vede arrivare una crisi.

Il dataset copre il 1980-2013(+) e segna sei periodi di recessione reali: 1980, 1981-1982, 1991, 2000-2001, dalla fine del 2007 a metà del 2009 (crisi finanziaria) e il 2020 (Covid-19). Oltre alle vendite mensili, porta PIL, disoccupazione, fiducia dei consumatori, spesa pubblicitaria, prezzo medio del veicolo e tipo di auto venduta.

**Quali grafici rispondono a cosa:**

- **Linea** per le vendite per anno: c'è una dimensione temporale continua di oltre 30 anni — la linea è l'unico modo per vedere tendenza e rotture a colpo d'occhio. Le barre qui sarebbero illeggibili.
- **Barre** per confrontare le vendite medie per tipo di veicolo in recessione vs. non recessione: sono categorie con nome, senza ordine temporale — il caso da manuale per le barre.
- **Pie** per la ripartizione della spesa pubblicitaria per tipo di veicolo: la domanda è "che proporzione del totale spetta a ogni categoria?", e con solo cinque categorie il pie resta leggibile.
- **Barre impilate per colore** per l'effetto della disoccupazione sulle vendite per tipo di auto: combina due dimensioni categoriche senza ricorrere a un grafico 3D.
- **Scatter/bubble** per collegare la fiducia dei consumatori o il prezzo con il volume di vendite: due variabili numeriche continue, il lavoro del punto di dispersione.

**La decisione che permette di prendere:** un direttore può vedere se le vendite di auto sportive sono calate più di quelle familiari durante una recessione specifica, e decidere se rafforzare la pubblicità sul segmento più resiliente la prossima volta che i segnali di PIL e disoccupazione anticipano una crisi.

## Caso 2 — Compagnie aeree: ritardi di volo per causa e per mese

**La domanda di business:** perché ritardano i voli, non solo quanto? Un ritardo per cattiva gestione dell'equipaggio (`carrier delay`) richiede una risposta diversa da uno causato da una tempesta (`weather delay`), dalla congestione del sistema aereo (`NAS delay`), dalla sicurezza (`security delay`) o perché l'aereo è arrivato in ritardo dal volo precedente (`late aircraft delay`).

**Quale grafico è stato usato e perché:** cinque pannelli, tutti a **linee**, una linea di colore per compagnia aerea (`color='Reporting_Airline'`). La dimensione temporale (il mese) e la necessità di vedere la tendenza stagionale escludono le barre — con dodici mesi e una dozzina di compagnie aeree, un grafico a barre raggruppate sarebbe un muro di colore illeggibile.

```python
carrier_fig = px.line(avg_car, x='Month', y='CarrierDelay',
                       color='Reporting_Airline',
                       title='Average carrier delay time (minutes) by airline')
```

**La decisione che permette di prendere:** un manager delle operazioni può vedere se il ritardo per "late aircraft" di una compagnia esplode nei mesi di maggior traffico, segnale di un problema di rotazione della flotta — mentre se il problema fosse il `weather delay`, la soluzione sarebbe una logistica di emergenza, non di programmazione.

## Caso 3 — Incendi boschivi in Australia

**La domanda di business:** quali regioni concentrano più attività di incendi? C'è stagionalità? È aumentata o diminuita negli anni? Un ente di monitoraggio ambientale raccoglie, dal 2005, dati giornalieri per ognuna delle sette regioni dell'Australia: area stimata dell'incendio, luminosità media del fuoco, potenza radiativa, confidenza della rilevazione e conteggio di pixel satellitari di vegetazione in fiamme.

**Quali grafici sono stati usati e perché:**

- **Linea** per l'area d'incendio nel tempo — dimensione temporale.
- **Barre (seaborn)** per confrontare la luminosità media del fuoco tra le sette regioni — categorie senza ordine temporale.
- **Pie** per la proporzione del conteggio di pixel per regione — poche categorie, ripartizione di un totale.
- **Istogramma** per la distribuzione della luminosità media del fuoco — qui la domanda non è più "come si confronta una regione con un'altra" ma "quanto è comune ogni livello di luminosità", terreno dell'istogramma.
- **Scatter** per collegare la potenza radiativa media con il livello di confidenza medio — due variabili numeriche continue.
- **Mappa con folium** che segna le sette regioni — l'unica delle tre storie con componente geografica centrale alla domanda ("dove succede questo?").

**La decisione che permette di prendere:** un analista ambientale può vedere quale regione concentra la maggiore proporzione di pixel in fiamme in un mese dato e dare priorità all'invio di risorse di spegnimento verso quella regione, invece di distribuirle equamente tra le sette.

## Il filo comune

Osserva cosa decide ogni grafico:

| Domanda | Tipo di grafico | Perché |
|---|---|---|
| Sale o scende nel tempo? | Linea | Dimensione temporale continua |
| Come si confrontano categorie con nome? | Barre | Categorie discrete, senza ordine temporale |
| Che proporzione di un totale ha ogni categoria? | Pie | Poche categorie, ripartizione di un totale |
| Quanto è comune ogni valore? | Istogramma | Distribuzione di una variabile numerica |
| Come si relazionano due variabili numeriche? | Scatter / bubble | Relazione tra due (o tre) variabili continue |
| Dove succede questo? | Mappa (folium) | Componente geografica centrale alla domanda |

Una dashboard reale quasi mai usa un solo tipo di grafico: combina diversi della tabella sopra a seconda della domanda a cui risponde ogni pannello. Prima di scegliere un tipo di grafico, chiediti sempre **chi lo guarderà e quale decisione deve prendere dopo averlo visto** — quella domanda, non la varietà di grafici che conosci, è ciò che guida il design di una dashboard utile.

## Preparare l'ambiente e caricare i dati

Usiamo lo stesso dataset `automobile-sales.csv` del progetto integrativo di automobili e recessione: vendite mensili tra il 1980 e il 2013+ con variabili economiche (PIL, disoccupazione, fiducia dei consumatori), pubblicità, prezzo e tipo di veicolo.

````python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

print('Librerie importate correttamente')
````

````python
import requests
import io

# URL del dataset delle vendite di automobili
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/d51iMGfp_t0QpO30Lym-dw/automobile-sales.csv"

response = requests.get(URL)
response.raise_for_status()

csv_content = io.StringIO(response.text)
df = pd.read_csv(csv_content)

print('Dati scaricati e letti in un dataframe!')
df.head()
````

## Passo 1 — Linea delle vendite per anno, con annotazioni di recessione

Iniziamo dalla domanda più basilare: come evolvono le vendite medie di automobili nel corso degli anni? Raggruppiamo per `Year` e calcoliamo la media di `Automobile_Sales`, poi annotiamo direttamente sul grafico i sei periodi di recessione storici così chi lo guarda può collegare direttamente i cali di vendite all'economia reale.

````python
df_line = df.groupby('Year')['Automobile_Sales'].mean()

plt.figure(figsize=(12, 6))
df_line.plot(kind='line')

plt.xticks(list(range(1980, int(df['Year'].max()) + 1)), rotation=75)
plt.xlabel('Anno')
plt.ylabel('Vendite medie di automobili')
plt.title('Vendite di automobili durante i periodi di recessione')

# Annotiamo i sei periodi di recessione storici direttamente sul grafico
plt.text(1980, df_line.max() * 0.05, '1980', rotation=90, fontsize=8, color='red')
plt.text(1981, df_line.max() * 0.05, '1981-82', rotation=90, fontsize=8, color='red')
plt.text(1991, df_line.max() * 0.05, '1991', rotation=90, fontsize=8, color='red')
plt.text(2000, df_line.max() * 0.05, '2000-01', rotation=90, fontsize=8, color='red')
plt.text(2007, df_line.max() * 0.05, '2007-09', rotation=90, fontsize=8, color='red')
plt.text(2020, df_line.max() * 0.05, '2020', rotation=90, fontsize=8, color='red')

plt.legend(['Vendite medie'])
plt.tight_layout()
plt.show()
````

**Lettura del grafico:** ogni linea verticale rossa segna l'inizio di una recessione reale. A colpo d'occhio si vede se le vendite calano proprio in quegli anni e con quale forza si riprendono dopo — questa è la domanda di business che ha motivato tutto il progetto.

## Passo 2 — Barre: vendite medie per tipo di veicolo, recessione vs. non recessione

Ora scendiamo di un livello: non solo "le vendite calano in recessione?" ma "quale tipo di auto regge meglio il calo?". Qui non c'è una dimensione temporale continua — ci sono cinque categorie con nome (`Vehicle_Type`) incrociate con una variabile binaria (`Recession`) — il caso da manuale per un grafico a barre raggruppate.

````python
grouped_df = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Recession', y='Automobile_Sales', hue='Vehicle_Type', data=grouped_df)

plt.xticks(ticks=[0, 1], labels=['Non recessione', 'Recessione'])
plt.xlabel('Periodo')
plt.ylabel('Vendite medie di automobili')
plt.title('Vendite per tipo di veicolo: recessione vs. non recessione')
plt.legend(title='Tipo di veicolo', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
````

**Lettura del grafico:** in generale tutte le categorie calano durante la recessione, ma non tutte calano allo stesso modo. Se `SuperMiniCar` cala meno di `Sports` o `ExecutiveCar`, questa è prova diretta che il segmento economico è più resiliente al ciclo economico — la stessa conclusione a cui arriva il manuale del capitolo.

## Passo 3 — Pie: spesa pubblicitaria per tipo di veicolo

Infine, su quale tipo di auto l'azienda concentra la sua pubblicità? Con solo cinque categorie e la domanda che è "che proporzione del totale spetta a ognuna", il grafico pie resta leggibile — con quindici categorie non lo sarebbe più, e lì converrebbe usare le barre.

````python
vt_expenditure = df.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()

plt.figure(figsize=(8, 6))
plt.pie(vt_expenditure.values, labels=vt_expenditure.index, autopct='%1.1f%%', startangle=90)
plt.title('Spesa pubblicitaria per tipo di veicolo')
plt.axis('equal')
plt.show()
````

**Lettura del grafico:** se il segmento economico (`SmallFamilyCar` o `SuperMiniCar`) concentra la fetta maggiore della spesa pubblicitaria, questo suggerisce che l'azienda punta già sul segmento più resiliente — coerente con quanto visto nel grafico a barre del passo precedente. I tre grafici insieme — linea, barre, pie — raccontano un'unica storia completa: quando calano le vendite, quale categoria resiste meglio, e dove l'azienda mette i suoi soldi per rafforzarla.
