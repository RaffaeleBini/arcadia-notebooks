def construye_target(df, periodo_futuro, ratio="LTC"):
    df = df.copy()
    df["future"] = df[f"{ratio}_close"].shift(-periodo_futuro)
    df["target"] = list(map(clasifica, df[f"{ratio}_close"], df["future"]))
    df.dropna(inplace=True)
