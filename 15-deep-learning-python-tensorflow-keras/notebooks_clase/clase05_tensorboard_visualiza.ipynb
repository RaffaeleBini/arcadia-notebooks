## Resumen para la clase

- TensorBoard visualiza en tiempo real la evolución del entrenamiento.
- El callback `TensorBoard(log_dir=...)` es lo único que necesitas añadir a `model.fit`.
- Da a cada modelo un nombre único (por ejemplo, con un timestamp).
- Compara `loss` contra `val_loss` para detectar overfitting.

```python
%load_ext tensorboard
```

```python
import pickle
import time
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.callbacks import TensorBoard

with open("X.pickle", "rb") as archivo:
    X = pickle.load(archivo)
with open("y.pickle", "rb") as archivo:
    y = pickle.load(archivo)
X = X / 255.0

nombre_modelo = f"cnn-gatos-perros-{int(time.time())}"
tensorboard_callback = TensorBoard(log_dir=f"logs/{nombre_modelo}")

model = Sequential([
    Conv2D(64, (3, 3), input_shape=X.shape[1:]),
    Activation("relu"),
    MaxPooling2D(pool_size=(2, 2)),
    Conv2D(64, (3, 3)),
    Activation("relu"),
    MaxPooling2D(pool_size=(2, 2)),
    Flatten(),
    Dense(64),
    Activation("relu"),
    Dense(1, activation="sigmoid"),
])

model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])

model.fit(
    X, y,
    batch_size=32,
    epochs=10,
    validation_split=0.1,
    callbacks=[tensorboard_callback],
)
```
