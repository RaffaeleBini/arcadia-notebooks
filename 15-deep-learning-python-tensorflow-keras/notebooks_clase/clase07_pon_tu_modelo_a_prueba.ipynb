## Resumen para la clase

- El accuracy de validación no garantiza que el modelo funcione con datos verdaderamente nuevos.
- Cualquier dato nuevo debe pasar por el **mismo preprocesamiento** que tus datos de entrenamiento.
- Probar con datos out-of-sample es una forma exigente de detectar overfitting.
- Un dataset de entrenamiento poco diverso limita la generalización del modelo.

```python
# Descarga un par de imagenes de prueba publicas (no forman parte del dataset de entrenamiento)
import tensorflow as tf

url_gato_prueba = "https://storage.googleapis.com/mledu-datasets/cats_and_dogs_filtered.zip"
# Reutilizamos el mismo dataset descargado en la clase 3: tomamos imagenes de su carpeta de validacion,
# que el modelo nunca vio durante el entrenamiento
import pathlib

ruta_zip = tf.keras.utils.get_file("cats_and_dogs_filtered.zip", origin=url_gato_prueba, extract=True)
directorio_validacion = pathlib.Path(ruta_zip).parent / "cats_and_dogs_filtered" / "validation"
print(list(directorio_validacion.iterdir()))
```

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

IMG_SIZE = 50
categorias = ["cats", "dogs"]


def preparar_imagen(ruta_archivo):
    array_imagen = cv2.imread(str(ruta_archivo), cv2.IMREAD_GRAYSCALE)
    array_redimensionado = cv2.resize(array_imagen, (IMG_SIZE, IMG_SIZE))
    return array_redimensionado.reshape(-1, IMG_SIZE, IMG_SIZE, 1) / 255.0


modelo = load_model("cnn_gatos_perros.keras")
```

```python
import os

carpeta_gatos_validacion = directorio_validacion / "cats"
primeras_imagenes = os.listdir(carpeta_gatos_validacion)[:5]

for nombre_archivo in primeras_imagenes:
    ruta = carpeta_gatos_validacion / nombre_archivo
    imagen_preparada = preparar_imagen(ruta)
    prediccion = modelo.predict(imagen_preparada, verbose=0)
    clase_predicha = categorias[int(round(prediccion[0][0]))]
    print(f"{nombre_archivo}: el modelo predice '{clase_predicha}' (etiqueta real: cats)")
```
