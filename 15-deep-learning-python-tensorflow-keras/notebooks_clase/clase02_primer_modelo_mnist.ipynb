## Resumen para la clase

- MNIST viene incluido en Keras: 70.000 imágenes de dígitos manuscritos de 28x28 píxeles.
- Normaliza siempre tus datos de entrada antes de entrenar.
- `Sequential` apila capas; `Flatten` convierte una imagen en un vector.
- Compilar = elegir optimizador, función de pérdida y métricas.
- `fit` entrena, `evaluate` mide sobre datos nuevos, `predict` genera predicciones.

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
print("Forma de x_train:", x_train.shape)
print("Forma de y_train:", y_train.shape)
```

```python
import matplotlib.pyplot as plt

plt.imshow(x_train[0], cmap="gray")
plt.title(f"Etiqueta: {y_train[0]}")
plt.show()
```

```python
# Normaliza: de [0, 255] a [0, 1]
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax"),
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

model.summary()
```

```python
model.fit(x_train, y_train, epochs=3)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Perdida en test: {loss:.4f}")
print(f"Precision en test: {accuracy:.4f}")
```

```python
import numpy as np

model.save("mi_primer_modelo.keras")

from tensorflow.keras.models import load_model
modelo_cargado = load_model("mi_primer_modelo.keras")

predicciones = modelo_cargado.predict(x_test)
digito_predicho = np.argmax(predicciones[0])
print(f"El modelo predice: {digito_predicho}")
print(f"La etiqueta real era: {y_test[0]}")
```

```python
from sklearn.metrics import confusion_matrix

y_pred = np.argmax(model.predict(x_test), axis=1)
print(confusion_matrix(y_test, y_pred))
```
