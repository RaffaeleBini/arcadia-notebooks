## Resumen para la clase

- Una RNN mantiene un estado oculto que se actualiza en cada paso de la secuencia.
- El vanishing gradient hace que las RNN simples "olviden" rápido en secuencias largas.
- LSTM usa puertas (olvido, entrada, salida) para mantener información relevante durante más tiempo.
- GRU simplifica LSTM con menos puertas.
- `return_sequences=True` es necesario cuando apilas una capa recurrente sobre otra.

```python
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

print("Forma de x_train (60000 secuencias de 28 pasos, 28 valores cada uno):", x_train.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

model = Sequential([
    LSTM(128, input_shape=(28, 28), return_sequences=True),
    Dropout(0.2),
    LSTM(128),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(10, activation="softmax"),
])

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
model.fit(x_train, y_train, epochs=3, validation_data=(x_test, y_test))
```
