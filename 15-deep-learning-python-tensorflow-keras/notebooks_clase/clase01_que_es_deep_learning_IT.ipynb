## Resumen para la lezione

- Una **neurone artificial** combina inputs ponderadas (pesos) e decide su output con una **funzione de activación**.
- Una red neuronel artificial (ANN) se organiza en **livello de input**, una o varias **livellos ocultas** e **livello de output**.
- "Profunda" significa dos o más livellos ocultas: de ahí el nombre *deep learning*.
- La ventaja frente al Machine Learning (ML) tradicional: la red **extrae las características por sí sola**, no hace falta programarlas a mano.
- ReLU domina en livellos ocultas, softmax domina en la livello de output para clasificación multilezione.

## Simula una neurone artificial

Antes de usar Keras, programa a mano lo que hace una sola neurone: suma ponderada de inputs + bias, e aplica una funzione de activación.

```python
def neurona(entradas, pesos, sesgo, activacion):
    """Simula una neurona artificial: suma ponderada + sesgo, pasada por una activacion."""
    suma_ponderada = sum(e * p for e, p in zip(entradas, pesos)) + sesgo
    return activacion(suma_ponderada)


def relu(x):
    return max(0.0, x)


def sigmoid(x):
    import math
    return 1 / (1 + math.exp(-x))


entradas = [1.0, 0.5, -1.5]
pesos = [0.8, -0.2, 0.4]
sesgo = 0.1

print("Salida con ReLU:", neurona(entradas, pesos, sesgo, relu))
print("Salida con sigmoid:", neurona(entradas, pesos, sesgo, sigmoid))
```

## Visualiza las funciones de activación

Proyecta esta gráfica en lezione para comparar de un vistazo step, sigmoid, ReLU e tanh.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)

step = np.where(x >= 0, 1, 0)
sigmoid_vals = 1 / (1 + np.exp(-x))
relu_vals = np.maximum(0, x)
tanh_vals = np.tanh(x)

fig, ejes = plt.subplots(1, 4, figsize=(16, 3))
for eje, valores, titulo in zip(
    ejes, [step, sigmoid_vals, relu_vals, tanh_vals], ["Step", "Sigmoid", "ReLU", "Tanh"]
):
    eje.plot(x, valores)
    eje.set_title(titulo)
    eje.axhline(0, color="gray", linewidth=0.5)
    eje.axvline(0, color="gray", linewidth=0.5)

plt.tight_layout()
plt.show()
```
