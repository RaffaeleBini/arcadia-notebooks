# 15.7 — Equazione di Poisson

**Prerequisiti:** 15.6 Equazione di Laplace, 14.6 Funzionali, 14.3 Spazi di Hilbert, 15.3 Spazi di Fourier.


## Perché ci interessa

L'equazione di Laplace (notebook 15.6) descrive l'equilibrio **senza sorgenti**. L'**equazione di Poisson** $-\Delta u=f$ aggiunge una sorgente $f$ (una densità di carica, una distribuzione di calore stazionaria con produzione interna): il problema classico di trovare $u$ diventa spesso troppo rigido per ammettere soluzioni classiche $C^2$ se $f$ non è sufficientemente regolare. La svolta, che apre la strada al Modulo 14 riapplicato qui, è la **formulazione variazionale**: riformulare il problema come la ricerca del minimo di un funzionale di energia in uno spazio di Hilbert (uno spazio di Sobolev, notebook 15.3), dove il teorema di rappresentazione di Riesz (notebook 14.6) garantisce l'esistenza e unicità della soluzione **debole**, anche quando la soluzione classica non esiste.

## Teoria, passo dopo passo

### Formulazione debole: moltiplicare e integrare per parti

Sia $\Omega\subseteq\mathbb R^n$ limitato, e cerchiamo $u$ con $-\Delta u=f$ in $\Omega$, $u=0$ su $\partial\Omega$ (condizione di Dirichlet omogenea). Moltiplichiamo per una funzione test $v$ (liscia, nulla sul bordo) e integriamo su $\Omega$: $-\int_\Omega(\Delta u)v\,dx=\int_\Omega fv\,dx$. Integrando per parti il primo membro (identità di Green, generalizzazione multidimensionale dell'integrazione per parti, Modulo 5/7, con il termine di bordo che si annulla essendo $v=0$ su $\partial\Omega$): $\int_\Omega\nabla u\cdot\nabla v\,dx=\int_\Omega fv\,dx$.

Questa identità **non richiede più** che $u$ sia due volte derivabile: basta che $\nabla u\in L^2(\Omega)$ (cioè $u$ nello spazio di Sobolev $H^1_0(\Omega)$, notebook 15.3, con il pedice $0$ che indica l'annullamento al bordo in senso debole). Diciamo che $u\in H^1_0(\Omega)$ è una **soluzione debole** se $\int_\Omega\nabla u\cdot\nabla v\,dx=\int_\Omega fv\,dx$ per **ogni** $v\in H^1_0(\Omega)$.

### Esistenza e unicità tramite Riesz

**Teorema (esistenza e unicità della soluzione debole).** Per ogni $f\in L^2(\Omega)$, esiste un'**unica** soluzione debole $u\in H^1_0(\Omega)$.

**Dimostrazione.** Su $H^1_0(\Omega)$ (uno spazio di Hilbert, con prodotto scalare $\langle u,v\rangle_{H^1_0}=\int_\Omega\nabla u\cdot\nabla v\,dx$ — una variante del notebook 15.3, valida grazie alla **disuguaglianza di Poincaré**, enunciata sotto, che garantisce che questa forma bilineare sia effettivamente un prodotto scalare completo), consideriamo il funzionale $\varphi(v)=\int_\Omega fv\,dx$. Per Cauchy-Schwarz (notebook 14.3) e la disuguaglianza di Poincaré: $|\varphi(v)|\le\|f\|_2\|v\|_2\le C\|f\|_2\|v\|_{H^1_0}$, quindi $\varphi$ è un funzionale lineare **continuo** su $H^1_0(\Omega)$ (notebook 14.6). Per il **teorema di rappresentazione di Riesz** (notebook 14.6), esiste un unico $u\in H^1_0(\Omega)$ con $\varphi(v)=\langle u,v\rangle_{H^1_0}=\int_\Omega\nabla u\cdot\nabla v\,dx$ per ogni $v$ — **esattamente** la definizione di soluzione debole. $\blacksquare$

**Il ruolo cruciale del Modulo 14.** Questa dimostrazione **non risolve mai esplicitamente** un'equazione differenziale: usa **solo** l'esistenza astratta garantita da Riesz. È un tipico esempio di come l'analisi funzionale trasformi un problema di esistenza di soluzioni di PDE in un problema puramente di struttura di spazi di Hilbert — un guadagno concettuale enorme rispetto ai metodi espliciti (Fourier, D'Alembert) dei notebook precedenti, al costo di ottenere una soluzione **debole**, non necessariamente classica.

### La disuguaglianza di Poincaré (enunciato)

**Teorema (Poincaré, enunciato).** Se $\Omega$ è limitato, esiste $C>0$ (dipendente solo da $\Omega$) tale che $\|v\|_2\le C\|\nabla v\|_2$ per ogni $v\in H^1_0(\Omega)$.

Questa disuguaglianza (dimostrabile con un argomento di compattezza che collega Rellich-Kondrachov, un raffinamento del teorema di immersione di Sobolev del notebook 15.3, alla contraddizione per assurdo) è ciò che garantisce che $\|\nabla v\|_2$ (che misura solo la "variazione" di $v$, non $v$ stessa) sia comunque una **norma completa** equivalente su $H^1_0(\Omega)$: senza l'annullamento al bordo ($v\in H^1_0$, non semplicemente $H^1$), una funzione costante avrebbe $\nabla v=0$ pur essendo $v\ne0$, rompendo la positività della norma.

### Regolarità: dalla soluzione debole a quella classica

**Teorema (regolarità ellittica, enunciato).** Se $f\in L^2(\Omega)$ e $\partial\Omega$ è sufficientemente regolare, la soluzione debole $u$ soddisfa automaticamente $u\in H^2(\Omega)$ (un guadagno di due derivate rispetto a $f\in L^2=H^0$). Se $f$ è più regolare (per esempio $f\in C^\infty$), $u$ è anch'essa $C^\infty$ all'interno di $\Omega$: la soluzione debole **coincide** con quella classica quando quest'ultima esiste, giustificando a posteriori l'intero approccio variazionale. Questo fenomeno di "guadagno di regolarità" è esattamente l'analogo, per equazioni alle derivate parziali generali, di quanto dimostrato esplicitamente per un caso particolare nell'esercizio 4 del notebook 15.3.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.sparse import diags
from scipy.sparse.linalg import spsolve

# Risolviamo numericamente -u'' = f su (0,1) con u(0)=u(1)=0 (versione 1D dell'equazione di Poisson)
# tramite differenze finite, per illustrare il concetto (la teoria e' in dimensione n generale,
# ma il caso 1D e' sufficiente per visualizzare esistenza/unicita' e il ruolo della formulazione debole)

N = 200
h = 1 / (N + 1)
x = np.linspace(0, 1, N+2)[1:-1]

def f(x):
    return np.sin(2*np.pi*x) * 4*np.pi**2  # scelto per avere soluzione esatta nota

# Matrice del laplaciano discreto (tridiagonale)
main_diag = 2 * np.ones(N) / h**2
off_diag = -1 * np.ones(N-1) / h**2
A = diags([off_diag, main_diag, off_diag], offsets=[-1, 0, 1]).tocsr()

b = f(x)
u_numerico = spsolve(A, b)

u_esatto = np.sin(2*np.pi*x)  # soluzione esatta per questa scelta di f

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x, u_numerico, 'bo', markersize=3, label='u numerica (differenze finite)')
ax.plot(x, u_esatto, 'r-', label='u esatta = sin(2*pi*x)')
ax.legend()
ax.set_title("Equazione di Poisson 1D: -u'' = f, u(0)=u(1)=0")
plt.tight_layout()
plt.show()

errore = np.max(np.abs(u_numerico - u_esatto))
print(f"Errore massimo tra soluzione numerica ed esatta: {errore:.6f}")
```

## Esercizio svolto e commentato

**Problema:** verifica la disuguaglianza di Poincaré (in dimensione $1$, come caso particolare illustrativo) per $\Omega=(0,1)$: dimostra che $\int_0^1v(x)^2\,dx\le\int_0^1v'(x)^2\,dx$ per ogni $v$ con $v(0)=0$ (una versione esplicita con $C=1$).

**Ragionamento:**

**Passo 1 — esprimiamo $v(x)$ come integrale della derivata.** Essendo $v(0)=0$: $v(x)=\int_0^xv'(t)\,dt$ (teorema fondamentale del calcolo, notebook 10.15).

**Passo 2 — applichiamo Cauchy-Schwarz all'integrale.** $v(x)^2=\Big(\int_0^xv'(t)\,dt\Big)^2=\Big(\int_0^x1\cdot v'(t)\,dt\Big)^2\le\Big(\int_0^x1^2\,dt\Big)\Big(\int_0^xv'(t)^2\,dt\Big)$ (Cauchy-Schwarz in $L^2([0,x])$, notebook 14.3, applicata alle funzioni $1$ e $v'$) $=x\int_0^xv'(t)^2\,dt\le x\int_0^1v'(t)^2\,dt$ (estendendo il dominio di integrazione, essendo l'integranda $\ge0$).

**Passo 3 — integriamo su $x\in[0,1]$.** $\int_0^1v(x)^2\,dx\le\int_0^1x\,dx\cdot\int_0^1v'(t)^2\,dt=\dfrac12\int_0^1v'(t)^2\,dt$ (essendo $\int_0^1x\,dx=1/2$, e il secondo fattore non dipende da $x$, esce dall'integrale).

**Conclusione:** $\int_0^1v^2\,dx\le\dfrac12\int_0^1(v')^2\,dx\le\int_0^1(v')^2\,dx$ (essendo $1/2\le1$): la disuguaglianza di Poincaré vale su $(0,1)$ con costante $C^2\le1/2$ (quindi $C\le1/\sqrt2$, anche più forte di quanto richiesto). **Punto metodologico:** questa tecnica — scrivere $v$ come integrale della sua derivata (sfruttando l'annullamento al bordo) e applicare Cauchy-Schwarz — è **esattamente** l'idea alla base della dimostrazione generale della disuguaglianza di Poincaré in dimensione superiore, dove l'integrazione avviene lungo segmenti radiali invece che sull'intervallo $[0,x]$.

```python
import numpy as np
from scipy import integrate

def v(x):
    return np.sin(np.pi * x)  # v(0)=0, v(1)=0 (test con una funzione concreta)

def v_prime(x):
    return np.pi * np.cos(np.pi * x)

norma_v_quadrato = integrate.quad(lambda x: v(x)**2, 0, 1)[0]
norma_vprime_quadrato = integrate.quad(lambda x: v_prime(x)**2, 0, 1)[0]

print(f"||v||_2^2 = {norma_v_quadrato:.5f}")
print(f"||v'||_2^2 = {norma_vprime_quadrato:.5f}")
print(f"Poincare (C=1) verificata: {norma_v_quadrato} <= {norma_vprime_quadrato}? {norma_v_quadrato <= norma_vprime_quadrato}")
print(f"Rapporto ||v||^2/||v'||^2 = {norma_v_quadrato/norma_vprime_quadrato:.5f} (atteso <= 1, e in questo caso esattamente 1/pi^2)")
```

4. *(più impegnativo)* Usando il risultato dell'esercizio 3 (il **principio di Dirichlet**: la soluzione debole minimizza l'energia), spiega perché questo fornisce un metodo alternativo — storicamente il primo, prima della teoria di Riesz — per dimostrare l'esistenza della soluzione: quali proprietà dello spazio $H^1_0(\Omega)$ servono per garantire che l'estremo inferiore di $J$ sia effettivamente **raggiunto** da qualche funzione (e non solo approssimato)?

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Prendendo $v=u$ nell'ipotesi: $\int_\Omega|\nabla u|^2\,dx=0$, cioè $\|\nabla u\|_2=0$. Per la disuguaglianza di Poincaré (teoria): $\|u\|_2\le C\|\nabla u\|_2=0$, quindi $\|u\|_2=0$, cioè $u=0$ (quasi ovunque, notebook 13.8).

**2.** Se $u_1,u_2\in H^1_0(\Omega)$ sono entrambe soluzioni deboli: $\int_\Omega\nabla u_1\cdot\nabla v\,dx=\int_\Omega fv\,dx=\int_\Omega\nabla u_2\cdot\nabla v\,dx$ per ogni $v$, quindi $\int_\Omega\nabla(u_1-u_2)\cdot\nabla v\,dx=0$ per ogni $v\in H^1_0(\Omega)$ (linearità dell'integrale e del gradiente). Per l'esercizio 1 (con $u=u_1-u_2$): $u_1-u_2=0$, cioè $u_1=u_2$.

**3.** $J(u+w)=\dfrac12\int_\Omega|\nabla(u+w)|^2\,dx-\int_\Omega f(u+w)\,dx=\dfrac12\int_\Omega|\nabla u|^2\,dx+\int_\Omega\nabla u\cdot\nabla w\,dx+\dfrac12\int_\Omega|\nabla w|^2\,dx-\int_\Omega fu\,dx-\int_\Omega fw\,dx$ (espandendo il quadrato della norma, bilinearità del prodotto scalare $H^1_0$). Riorganizzando: $=J(u)+\Big(\int_\Omega\nabla u\cdot\nabla w\,dx-\int_\Omega fw\,dx\Big)+\dfrac12\|\nabla w\|_2^2$. Il termine tra parentesi è **esattamente $0$** (essendo $u$ soluzione debole, con $v=w$ nella formulazione debole). Dunque $J(u+w)=J(u)+\dfrac12\|\nabla w\|_2^2\ge J(u)$ (essendo l'ultimo termine sempre $\ge0$), con uguaglianza solo se $\|\nabla w\|_2=0$, cioè (per Poincaré, esercizio 1) $w=0$. Dunque $u$ è l'unico minimo di $J$.

**4.** Per garantire che l'estremo inferiore di $J$ sia raggiunto, serve un argomento di **compattezza debole**: si prende una successione minimizzante $(v_n)$ con $J(v_n)\to\inf J$; usando che $J$ controlla $\|\nabla v_n\|_2$ (quindi, per Poincaré, anche $\|v_n\|_{H^1_0}$), la successione è **limitata** in $H^1_0(\Omega)$; si estrae una sottosuccessione debolmente convergente (un concetto — la convergenza debole, distinta da quella in norma — che generalizza ulteriormente gli strumenti degli spazi di Hilbert, tipicamente trattato subito dopo il materiale di questo modulo); infine si usa la **semicontinuità inferiore debole** della norma (una conseguenza della convessità di $\|\cdot\|^2$) per dedurre che il limite debole $u$ soddisfa $J(u)\le\liminf J(v_n)=\inf J$, quindi $J(u)=\inf J$: il minimo è raggiunto. Questo metodo, detto **metodo diretto del calcolo delle variazioni**, richiede dunque che $H^1_0(\Omega)$ sia uno spazio di Hilbert (per la compattezza debole della palla limitata, un fatto sempre vero in spazi di Hilbert, a differenza della compattezza forte che fallisce in dimensione infinita per il teorema di Riesz del notebook 14.11) — un'illustrazione ulteriore di come le proprietà strutturali del Modulo 14 siano il fondamento tecnico di questi argomenti di esistenza.

## Riepilogo finale

- L'**equazione di Poisson** $-\Delta u=f$, riformulata **debolmente** tramite integrazione per parti, diventa un'identità su $H^1_0(\Omega)$ che non richiede più derivate seconde classiche.
- Il **teorema di rappresentazione di Riesz** (notebook 14.6), applicato al funzionale $v\mapsto\int fv$, garantisce esistenza e unicità della soluzione debole — un argomento di puro Modulo 14, senza costruzione esplicita.
- La **disuguaglianza di Poincaré** garantisce che $\|\nabla\cdot\|_2$ sia una norma completa equivalente su $H^1_0(\Omega)$, essenziale per applicare Riesz.
- Il **principio di Dirichlet** interpreta la soluzione debole come il minimizzante di un funzionale di energia: un punto di vista alternativo, storicamente precedente, che anticipa il **metodo diretto del calcolo delle variazioni**.
