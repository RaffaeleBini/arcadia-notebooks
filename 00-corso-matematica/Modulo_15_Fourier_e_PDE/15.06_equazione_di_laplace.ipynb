Le equazioni del calore (notebook 15.4) e delle onde (notebook 15.5) descrivono **evoluzione nel tempo**. L'**equazione di Laplace** $\Delta u=0$ (dove $\Delta u=\partial_x^2u+\partial_y^2u$ è il **laplaciano**, la traccia della matrice Hessiana, notebook 7.9-7.11, ottenibile anche come divergenza del gradiente, notebook 7.21) descrive invece uno **stato di equilibrio**: la temperatura di una piastra dopo un tempo lungo, il potenziale elettrostatico in una regione senza cariche. Le funzioni che risolvono $\Delta u=0$, dette **armoniche**, hanno proprietà notevoli — la proprietà della media, il principio del massimo — che le rendono tra gli oggetti più studiati e "rigidi" dell'analisi, e forniscono il caso modello per l'intera classe delle equazioni **ellittiche** (notebook 15.8).

## Teoria, passo dopo passo

### Funzioni armoniche e la proprietà della media

Una funzione $u\in C^2(\Omega)$ (due volte derivabile con continuità, Modulo 7) è **armonica** se $\Delta u=0$ su un aperto $\Omega\subseteq\mathbb R^2$ (o $\mathbb R^n$).

**Teorema (proprietà della media).** Se $u$ è armonica su $\Omega$ e $\overline B(x_0,r)\subseteq\Omega$, allora $u(x_0)=\dfrac1{2\pi r}\oint_{\partial B(x_0,r)}u\,ds$ (la media di $u$ sulla circonferenza di raggio $r$ centrata in $x_0$ è esattamente $u(x_0)$).

**Dimostrazione (schema, con il teorema della divergenza).** Definiamo $\phi(r)=\dfrac1{2\pi r}\oint_{\partial B(x_0,r)}u\,ds$. Derivando rispetto a $r$ (parametrizzando il cerchio e usando la regola di derivazione sotto il segno di integrale) e applicando il **teorema della divergenza** (notebook 7.24): $\phi'(r)=\dfrac1{2\pi r}\oint_{\partial B(x_0,r)}\partial_\nu u\,ds=\dfrac1{2\pi r}\iint_{B(x_0,r)}\Delta u\,dA=0$ (usando $\Delta u=0$, l'ipotesi di armonicità). Dunque $\phi$ è **costante**; per continuità, $\phi(r)\to u(x_0)$ quando $r\to0^+$ (la media su cerchi sempre più piccoli tende al valore centrale, per continuità di $u$): dunque $\phi(r)=u(x_0)$ per ogni $r$. $\blacksquare$

### Il principio del massimo

**Teorema (principio del massimo forte).** Se $u$ è armonica su un aperto **connesso** e limitato $\Omega$ (notebook 12.5), continua su $\overline\Omega$, e raggiunge il suo massimo in un punto **interno** $x_0\in\Omega$, allora $u$ è **costante** su $\Omega$.

**Dimostrazione (schema).** Per la proprietà della media, $u(x_0)$ è la media di $u$ su ogni cerchio abbastanza piccolo attorno a $x_0$; essendo $u(x_0)$ il massimo, e la media di valori $\le u(x_0)$ che eguaglia $u(x_0)$ stesso, $u$ deve essere **identicamente uguale** a $u(x_0)$ su quel cerchio (se fosse strettamente minore anche in un solo punto, per continuità lo sarebbe su un piccolo arco, abbassando la media sotto $u(x_0)$, contraddizione). Ripetendo questo argomento e usando la connessione di $\Omega$ (un argomento di "insieme aperto e chiuso allo stesso tempo", analogo alla dimostrazione del notebook 12.5 sulla connessione), l'insieme dove $u=u(x_0)$ è tutto $\Omega$. $\blacksquare$

**Corollario (principio del massimo debole).** $\max_{\overline\Omega}u=\max_{\partial\Omega}u$: il massimo (e per linearità, applicando a $-u$, anche il minimo) di una funzione armonica su un dominio limitato è sempre raggiunto sul **bordo**, mai (esclusivamente) all'interno a meno che $u$ sia costante.

### Unicità della soluzione del problema di Dirichlet

**Corollario (unicità).** Il problema $\Delta u=0$ su $\Omega$, $u=g$ su $\partial\Omega$ (dato $g$ fissato sul bordo) ha **al più una** soluzione continua su $\overline\Omega$.

**Dimostrazione.** Se $u_1,u_2$ sono entrambe soluzioni, $w=u_1-u_2$ è armonica (per linearità del laplaciano) con $w=0$ su $\partial\Omega$. Per il principio del massimo debole, $\max_{\overline\Omega}w=\max_{\partial\Omega}w=0$ e (applicato a $-w$) $\min_{\overline\Omega}w=0$: dunque $w\equiv0$, cioè $u_1=u_2$. $\blacksquare$ Questo è un tipico argomento di unicità "per differenza", identico nello spirito a quello usato per l'unicità delle soluzioni di ODE lineari (Modulo 8).

### Soluzione sul disco tramite serie di Fourier

**Teorema (formula di Poisson, costruzione tramite Fourier).** Sul disco unitario $\mathbb D=\{(x,y):x^2+y^2<1\}$, usando le coordinate polari $u(\rho,\theta)$ (Modulo 2) e cercando soluzioni della forma $u(\rho,\theta)=\sum_n\rho^{|n|}c_ne^{in\theta}$ (una serie di Fourier in $\theta$, notebook 15.1, con coefficienti che dipendono da $\rho$): sostituendo nell'equazione di Laplace in coordinate polari ($\Delta u=\partial_\rho^2u+\frac1\rho\partial_\rho u+\frac1{\rho^2}\partial_\theta^2u=0$) si trova che $\rho^{|n|}$ è **esattamente** la scelta che rende ogni termine della serie armonico separatamente (verifica diretta per sostituzione, un'equazione di Eulero in $\rho$). Imponendo il dato al bordo $u(1,\theta)=g(\theta)$: $c_n$ sono i coefficienti di Fourier di $g$ (notebook 15.1). Questo fornisce una soluzione esplicita del problema di Dirichlet sul disco, con unicità garantita dal corollario sopra.

```python
import numpy as np
import matplotlib.pyplot as plt

# Verifica numerica della proprieta' della media per una funzione armonica: u(x,y) = x^2 - y^2
def u(x, y):
    return x**2 - y**2  # armonica: d^2u/dx^2 + d^2u/dy^2 = 2 - 2 = 0

x0, y0 = 0.5, 0.3
u_x0 = u(x0, y0)

for r in [0.1, 0.3, 0.5]:
    theta = np.linspace(0, 2*np.pi, 1000)
    x_cerchio = x0 + r*np.cos(theta)
    y_cerchio = y0 + r*np.sin(theta)
    media = np.trapz(u(x_cerchio, y_cerchio), theta) / (2*np.pi)
    print(f"r={r}: media di u sul cerchio = {media:.6f}, u(x0,y0) = {u_x0:.6f}")

print("\nLa media coincide sempre con u(x0,y0), verificando la proprieta' della media")

# Verifica che u NON e' armonica se aggiungiamo un termine che rompe Delta u = 0
def v(x, y):
    return x**2 + y**2  # NON armonica: Delta v = 2+2 = 4 != 0

for r in [0.1, 0.3]:
    theta = np.linspace(0, 2*np.pi, 1000)
    x_cerchio = x0 + r*np.cos(theta)
    y_cerchio = y0 + r*np.sin(theta)
    media_v = np.trapz(v(x_cerchio, y_cerchio), theta) / (2*np.pi)
    print(f"r={r}: media di v (non armonica) = {media_v:.6f}, v(x0,y0) = {v(x0,y0):.6f} (diversi!)")
```

## Esercizio svolto e commentato

**Problema:** risolvi il problema di Dirichlet sul disco unitario con dato al bordo $g(\theta)=3\cos\theta+2\sin(2\theta)$, usando la formula della teoria.

**Ragionamento:**

**Passo 1 — troviamo i coefficienti di Fourier di $g$.** Scrivendo in notazione reale (invece che complessa, per collegarsi direttamente al notebook 15.1): $g(\theta)=3\cos\theta+2\sin(2\theta)$ è **già** scritta come somma finita di armoniche di Fourier, quindi i suoi coefficienti sono immediati: $a_1=3$ (coefficiente di $\cos\theta$), $b_2=2$ (coefficiente di $\sin(2\theta)$), tutti gli altri nulli.

**Passo 2 — applichiamo la formula con $\rho^{|n|}$.** Per l'armonica di ordine $1$ ($\cos\theta$), il fattore radiale è $\rho^1=\rho$; per quella di ordine $2$ ($\sin(2\theta)$), il fattore è $\rho^2$. Dunque: $u(\rho,\theta)=3\rho\cos\theta+2\rho^2\sin(2\theta)$.

**Passo 3 — verifichiamo che soddisfa il dato al bordo.** Per $\rho=1$: $u(1,\theta)=3\cos\theta+2\sin(2\theta)=g(\theta)$ ✓.

**Passo 4 — verifichiamo direttamente l'armonicità (controllo di consistenza).** Convertendo in coordinate cartesiane: $\rho\cos\theta=x$ (armonica banalmente, essendo lineare: $\Delta x=0$). $\rho^2\sin(2\theta)=\rho^2\cdot2\sin\theta\cos\theta=2(\rho\sin\theta)(\rho\cos\theta)=2xy$: verifichiamo $\Delta(2xy)=\partial_x^2(2xy)+\partial_y^2(2xy)=0+0=0$ (essendo $2xy$ lineare in ciascuna variabile separatamente, le derivate seconde pure si annullano).

**Conclusione:** $u(x,y)=3x+4xy$ (riscrivendo $2\rho^2\sin(2\theta)=4xy$) è l'unica soluzione (per il corollario di unicità) del problema di Dirichlet con questo dato al bordo. **Punto metodologico:** quando il dato al bordo è già una combinazione finita di armoniche trigonometriche, la soluzione si ottiene immediatamente moltiplicando ogni termine per la potenza $\rho^{|n|}$ corrispondente — senza bisogno di calcolare esplicitamente una serie infinita, un caso particolarmente semplice ma istruttivo della formula generale.

```python
import sympy as sp

x, y = sp.symbols('x y', real=True)
u = 3*x + 4*x*y

laplaciano = sp.diff(u, x, 2) + sp.diff(u, y, 2)
print("Laplaciano di u = 3x + 4xy:", laplaciano)

# Verifica del dato al bordo in coordinate polari
theta = sp.symbols('theta', real=True)
rho = 1
u_bordo = u.subs({x: rho*sp.cos(theta), y: rho*sp.sin(theta)})
u_bordo_semplificato = sp.simplify(sp.expand_trig(u_bordo))
print("u(1,theta) =", u_bordo_semplificato)
print("Atteso: 3*cos(theta) + 2*sin(2*theta)")
atteso = 3*sp.cos(theta) + 2*sp.sin(2*theta)
print("Coincidono?", sp.simplify(u_bordo - atteso) == 0)
```

## Riepilogo finale

- L'**equazione di Laplace** $\Delta u=0$ descrive stati di equilibrio; le sue soluzioni, dette **armoniche**, soddisfano la notevole **proprietà della media**: il valore in un punto è la media sui cerchi centrati in quel punto.
- Il **principio del massimo** garantisce che una funzione armonica non costante non può avere massimi (né minimi) interni: il valore estremo è sempre sul bordo del dominio.
- Questo principio implica direttamente l'**unicità** della soluzione del problema di Dirichlet: due soluzioni con lo stesso dato al bordo devono coincidere.
- Sul disco, la soluzione si costruisce esplicitamente tramite serie di Fourier del dato al bordo, moltiplicando ogni armonica $e^{in\theta}$ per il fattore radiale $\rho^{|n|}$ — un'applicazione diretta della teoria del notebook 15.1.
