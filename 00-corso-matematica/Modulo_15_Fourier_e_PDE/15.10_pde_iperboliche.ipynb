## Perché ci interessa

Il notebook 15.9 ha risolto l'equazione del calore su un dominio limitato tramite la base spettrale del laplaciano, trovando un decadimento esponenziale verso l'equilibrio. Questo notebook, capstone del Modulo 15 e dell'intero percorso di analisi, applica la **stessa base spettrale** all'equazione delle **onde** (generalizzando il notebook 15.5 a domini limitati), scoprendo un comportamento radicalmente diverso: **nessun decadimento**, ma **conservazione perpetua dell'energia**, esattamente come previsto dal contrasto qualitativo calore/onde già osservato su $\mathbb R$. Questo chiude il cerchio dell'intero corso: la stessa struttura astratta (teorema spettrale, spazi di Hilbert) produce comportamenti fisici opposti a seconda della natura dell'equazione — dissipativa (parabolica) o conservativa (iperbolica) — dimostrando il potere unificante e al contempo discriminante degli strumenti costruiti nei Moduli 12-15.

## Teoria, passo dopo passo

### Il problema iperbolico astratto

Sullo stesso dominio $\Omega$ e la stessa base spettrale $\{\phi_k,\lambda_k\}$ del notebook 15.9, consideriamo $\partial_t^2u=\Delta u$ su $\Omega\times(0,\infty)$, $u=0$ su $\partial\Omega$, con **due** dati iniziali (essendo del secondo ordine in $t$, come nel notebook 15.5): $u(\cdot,0)=f$, $\partial_tu(\cdot,0)=g$.

### Separazione di variabili astratta

Sviluppando $f=\sum_ka_k\phi_k$, $g=\sum_kb_k\phi_k$ e cercando $u(x,t)=\sum_kc_k(t)\phi_k(x)$: sostituendo, come nel notebook 15.9, si ottiene $c_k''(t)=-\lambda_kc_k(t)$ per ogni $k$ — un'**ODE lineare del secondo ordine oscillatoria** (notebook 8.4), identica nella forma a quella incontrata nel notebook 15.5 per la trasformata di Fourier, ma ora con $\lambda_k$ discreto al posto di $4\pi^2\xi^2$ continuo. La soluzione generale è $c_k(t)=a_k\cos(\sqrt{\lambda_k}t)+\dfrac{b_k}{\sqrt{\lambda_k}}\sin(\sqrt{\lambda_k}t)$ (imponendo $c_k(0)=a_k$, $c_k'(0)=b_k$).

**Formula della soluzione:** $u(x,t)=\sum_k\Big[a_k\cos(\sqrt{\lambda_k}t)+\dfrac{b_k}{\sqrt{\lambda_k}}\sin(\sqrt{\lambda_k}t)\Big]\phi_k(x)$ — l'analogo discreto della formula del notebook 15.5, con $\sqrt{\lambda_k}$ al posto di $2\pi\xi$.

### Conservazione dell'energia (nel linguaggio spettrale)

**Teorema.** L'energia $E(t)=\dfrac12\|\partial_tu(\cdot,t)\|_2^2+\dfrac12\|\nabla u(\cdot,t)\|_2^2$ è **costante** nel tempo.

**Dimostrazione.** $\partial_tu(x,t)=\sum_k\sqrt{\lambda_k}\Big[-a_k\sin(\sqrt{\lambda_k}t)+\dfrac{b_k}{\sqrt{\lambda_k}}\cos(\sqrt{\lambda_k}t)\Big]\phi_k(x)$. Per Pitagora generalizzato (ortonormalità dei $\phi_k$, notebook 14.3): $\|\partial_tu(\cdot,t)\|_2^2=\sum_k\lambda_k\Big[a_k\sin(\sqrt{\lambda_k}t)-\dfrac{b_k}{\sqrt{\lambda_k}}\cos(\sqrt{\lambda_k}t)\Big]^2$ (con un cambio di segno interno che non altera il quadrato). Analogamente, usando $\|\nabla\phi_k\|_2^2=\lambda_k$ (dal notebook 15.9, esercizio 3): $\|\nabla u(\cdot,t)\|_2^2=\sum_k\lambda_k\Big[a_k\cos(\sqrt{\lambda_k}t)+\dfrac{b_k}{\sqrt{\lambda_k}}\sin(\sqrt{\lambda_k}t)\Big]^2$. Sommando le due espressioni: per ogni $k$, il contributo è $\lambda_k\big[(\ldots\sin-\ldots\cos)^2+(\ldots\cos+\ldots\sin)^2\big]$, che si espande (usando $\sin^2+\cos^2=1$, Modulo 2, e la cancellazione dei termini incrociati $\pm2\sin\cos\cdot(\text{stesso prodotto})$) a $\lambda_k(a_k^2+b_k^2/\lambda_k)=\lambda_ka_k^2+b_k^2$, **indipendente da $t$**. Dunque $2E(t)=\sum_k(\lambda_ka_k^2+b_k^2)$, una costante. $\blacksquare$

**Contrasto netto con il notebook 15.9.** Mentre per il calore ogni "modo" $c_k(t)=c_ke^{-\lambda_kt}$ decade esponenzialmente, qui ogni modo **oscilla indefinitamente** senza perdere ampiezza: $c_k(t)^2+\big(c_k'(t)/\sqrt{\lambda_k}\big)^2$ è costante per ogni singolo $k$ (verifica diretta, identità di Pitagora per ciascun'onda armonica separatamente) — la stessa struttura astratta (base spettrale) produce **dissipazione** per l'equazione parabolica e **conservazione perpetua** per quella iperbolica, una dicotomia che riflette esattamente il segno della derivata temporale nell'equazione (primo ordine, con decadimento, contro secondo ordine, oscillatorio).

### Assenza di effetto regolarizzante (anche su domini limitati)

**Osservazione.** A differenza del notebook 15.9 (dove $u(\cdot,t)\in H^s$ per ogni $s$, per $t>0$, indipendentemente dal dato), qui il coefficiente $\lambda_k^s\big[a_k^2\cos^2(\sqrt{\lambda_k}t)+\cdots\big]$ **non decade** in $k$ più di quanto già garantito dal dato iniziale stesso (essendo le funzioni $\cos,\sin$ limitate ma non decrescenti): la regolarità $H^s$ della soluzione al tempo $t$ è **determinata dalla regolarità dei dati iniziali** $(f,g)$, senza guadagno né perdita — l'ultima conferma, nel linguaggio spettrale astratto, dello stesso fenomeno "nessun effetto regolarizzante" già osservato esplicitamente con la formula di D'Alembert nel notebook 15.5.

```python
import numpy as np
import matplotlib.pyplot as plt

def phi_k(x, k):
    return np.sqrt(2) * np.sin(k*np.pi*x)

def lambda_k(k):
    return (k*np.pi)**2

def f(x):
    return x*(1-x)
def g(x):
    return np.zeros_like(x)

x = np.linspace(0, 1, 500)
K = 20
a_k = [np.trapz(f(x)*phi_k(x,k), x) for k in range(1, K+1)]
b_k = [np.trapz(g(x)*phi_k(x,k), x) for k in range(1, K+1)]

def u_onde(x, t, a_k, b_k, K):
    s = np.zeros_like(x)
    for k in range(1, K+1):
        lk = lambda_k(k)
        sk = np.sqrt(lk)
        s += (a_k[k-1]*np.cos(sk*t) + b_k[k-1]/sk*np.sin(sk*t)) * phi_k(x, k)
    return s

fig, ax = plt.subplots(figsize=(8,5))
for t in [0, 0.05, 0.15, 0.3]:
    ax.plot(x, u_onde(x, t, a_k, b_k, K), label=f"t={t}")
ax.legend()
ax.set_title("Equazione delle onde su (0,1): oscilla, NON decade")
plt.tight_layout()
plt.show()

# Verifica della conservazione dell'energia
tempi = np.linspace(0, 2, 50)
energie = []
for t in tempi:
    u_t = u_onde(x, t, a_k, b_k, K)
    # approssimazione numerica di d/dt u tramite differenza finita
    dt = 1e-5
    u_t_dt = u_onde(x, t+dt, a_k, b_k, K)
    dudt = (u_t_dt - u_t) / dt
    dudx = np.gradient(u_t, x)
    E = 0.5*np.trapz(dudt**2, x) + 0.5*np.trapz(dudx**2, x)
    energie.append(E)

fig, ax = plt.subplots(figsize=(7,4))
ax.plot(tempi, energie)
ax.set_title("Energia E(t): approssimativamente costante nel tempo")
ax.set_ylim(min(energie)*0.9, max(energie)*1.1)
plt.tight_layout()
plt.show()
print(f"Energia: min={min(energie):.5f}, max={max(energie):.5f} (variazione dovuta solo ad approssimazione numerica)")
```

## Esercizio svolto e commentato

**Problema:** su $(0,\pi)$, risolvi $\partial_t^2u=\partial_x^2u$, $u(0,t)=u(\pi,t)=0$, con $u(x,0)=\sin(x)$, $\partial_tu(x,0)=3\sin(2x)$, e verifica esplicitamente la conservazione dell'energia confrontando $E(0)$ ed $E(1)$.

**Ragionamento:**

**Passo 1 — identifichiamo i coefficienti.** $a_1=1$ ($\lambda_1=1$), $b_2=3$ ($\lambda_2=4$), tutti gli altri nulli.

**Passo 2 — scriviamo la soluzione.** $u(x,t)=\cos(t)\sin(x)+\dfrac3{\sqrt4}\sin(2t)\sin(2x)=\cos(t)\sin(x)+\dfrac32\sin(2t)\sin(2x)$ (usando $\sqrt{\lambda_2}=\sqrt4=2$).

**Passo 3 — calcoliamo $E(0)$.** Al tempo $0$: $u(x,0)=\sin(x)$ (verifica: $\cos(0)=1$, $\sin(0)=0$ ✓), quindi $\nabla u(\cdot,0)=\cos(x)$, con $\|\nabla u(\cdot,0)\|_2^2=\int_0^\pi\cos^2(x)\,dx=\pi/2$. $\partial_tu(x,0)=3\sin(2x)$ (per costruzione), con $\|\partial_tu(\cdot,0)\|_2^2=9\int_0^\pi\sin^2(2x)\,dx=9\cdot\pi/2$. Dunque $E(0)=\dfrac12\Big(\dfrac{9\pi}2+\dfrac\pi2\Big)=\dfrac12\cdot5\pi=\dfrac{5\pi}2$.

**Passo 4 — usiamo la formula generale della teoria invece di ricalcolare a $t=1$ da zero.** $2E(t)=\sum_k(\lambda_ka_k^2+b_k^2)=\lambda_1a_1^2+b_2^2$ (essendo solo $k=1,2$ non nulli, con $a_1=1,b_2=3$, e $b_1=0,a_2=0$) $=1\cdot1^2+3^2=1+9=10$. Dunque $E(t)=5$ per **ogni** $t$, incluso $t=1$.

**Verifica di coerenza:** $E(0)=5\pi/2\approx7.85$ dal calcolo diretto (passo 3), ma la formula generale dà $E=5$ — **discrepanza dovuta alla normalizzazione**: le autofunzioni $\sin(kx)$ usate qui **non sono normalizzate** in $L^2(0,\pi)$ (hanno $\|\sin(kx)\|_2^2=\pi/2$, non $1$). Ripetendo il calcolo con $\phi_k(x)=\sqrt{2/\pi}\sin(kx)$ (normalizzate) e i coefficienti corrispondenti $\tilde a_1=\sqrt{\pi/2}$, $\tilde b_2=3\sqrt{\pi/2}$: $2E=\lambda_1\tilde a_1^2+\tilde b_2^2=1\cdot\pi/2+9\cdot\pi/2=5\pi$, dando $E=5\pi/2$ — **ora coerente** con il calcolo diretto del passo 3.

**Conclusione:** $E(t)=5\pi/2$ per ogni $t$ (una volta usata la normalizzazione corretta): l'energia è conservata, verificato sia dalla formula spettrale generale sia (implicitamente, per costruzione) dal calcolo diretto. **Lezione metodologica:** questo esercizio è un utile promemoria che le formule "eleganti" della teoria astratta richiedono di **tracciare con cura le normalizzazioni** quando si passa a basi esplicite non normalizzate come $\sin(kx)$ su $(0,\pi)$.

```python
import numpy as np
from scipy import integrate

def u(x, t):
    return np.cos(t)*np.sin(x) + 1.5*np.sin(2*t)*np.sin(2*x)

def dudt(x, t, dt=1e-6):
    return (u(x, t+dt) - u(x, t-dt)) / (2*dt)

def dudx_func(x, t, dx=1e-6):
    return (u(x+dx, t) - u(x-dx, t)) / (2*dx)

x = np.linspace(0.001, np.pi-0.001, 1000)

for t in [0, 0.5, 1, 2]:
    ut = dudt(x, t)
    ux = dudx_func(x, t)
    E = 0.5*integrate.trapezoid(ut**2, x) + 0.5*integrate.trapezoid(ux**2, x)
    print(f"t={t}: E(t) = {E:.5f} (atteso: 5*pi/2 = {5*np.pi/2:.5f})")
```

Per le onde, l'ODE $c_k''=-\lambda_kc_k$ è del **secondo ordine oscillatorio** (radici caratteristiche immaginarie pure, non reali negative), con soluzione $c_k(t)=a_k\cos(\sqrt{\lambda_k}t)+\ldots\sin(\sqrt{\lambda_k}t)$: un'oscillazione di ampiezza costante, mai un decadimento. **4.** $2E(0)=\sum_k(\lambda_ka_k^2+b_k^2)$, dove $a_k=\langle f,\phi_k\rangle$, $b_k=\langle g,\phi_k\rangle$. Il termine $\sum_kb_k^2=\|g\|_2^2<\infty$ (Parseval, essendo $g\in L^2$ per ipotesi). Il termine $\sum_k\lambda_ka_k^2=\|\nabla f\|_2^2$ (per l'identità $\|\nabla\phi_k\|_2^2=\lambda_k$ del notebook 15.9, e Parseval applicato a $\nabla f=\sum_ka_k\nabla\phi_k$, essendo i $\nabla\phi_k$ ortogonali con questa norma): finito, essendo $f\in H^1_0(\Omega)$ per ipotesi (quindi $\nabla f\in L^2(\Omega)$, per la caratterizzazione degli spazi di Sobolev, notebook 15.3). Per il teorema di conservazione (teoria), $E(t)=E(0)<\infty$ per ogni $t\ge0$: la soluzione ha energia finita per sempre, un risultato di buona posizione (esistenza di soluzioni fisicamente sensate) per l'equazione delle onde su domini limitati.

## Riepilogo finale

- Il problema delle onde su un dominio limitato, risolto con la stessa base spettrale del notebook 15.9, dà $c_k''(t)=-\lambda_kc_k(t)$: un'ODE oscillatoria, non dissipativa.
- L'**energia** $E(t)=\frac12\|\partial_tu\|_2^2+\frac12\|\nabla u\|_2^2$ è **esattamente conservata** nel tempo — il contrasto netto con il decadimento esponenziale dell'equazione parabolica.
- Nessun effetto regolarizzante: la regolarità della soluzione riflette esattamente quella dei dati iniziali, senza guadagno né perdita, in ogni istante.
- Questo chiude il Modulo 15 (e con esso l'intero blocco di analisi funzionale/PDE): la stessa base spettrale astratta (teorema spettrale, notebook 14.12) produce comportamenti fisici opposti — dissipazione per il calore, conservazione per le onde — a seconda dell'ordine della derivata temporale, mostrando come gli strumenti unificanti dell'analisi funzionale sappiano comunque discriminare con precisione fenomeni fisici qualitativamente diversi.
