# 15.2 — Trasformata di Fourier (rigorosa)

**Prerequisiti:** 15.1 Serie di Fourier (rigorose), 13.4 Integrale di Lebesgue, 14.3 Spazi di Hilbert, 8.10 Trasformata di Fourier (introduzione).


## Perché ci interessa

La serie di Fourier (notebook 15.1) scompone una funzione **periodica** in una somma discreta di frequenze $n=0,1,2,\ldots$ Una funzione definita su tutto $\mathbb R$, non periodica, richiede l'analogo continuo: invece di una somma su frequenze intere, un **integrale** su frequenze reali $\xi\in\mathbb R$. Questo notebook costruisce rigorosamente la trasformata di Fourier, prima su $L^1$ (dove l'integrale che la definisce converge in modo classico, notebook 13.4) e poi, tramite un argomento di densità, su $L^2$, dove diventa un'**isometria** — il fatto più importante, che generalizza l'identità di Parseval del notebook 15.1 a un contesto continuo.

## Teoria, passo dopo passo

### Definizione su $L^1(\mathbb R)$

Per $f\in L^1(\mathbb R)$ (notebook 13.8, con $\int_{\mathbb R}|f|\,dx<\infty$), la **trasformata di Fourier** è $\hat f(\xi)=\int_{\mathbb R}f(x)e^{-2\pi ix\xi}\,dx$ (integrale di Lebesgue, notebook 13.4, ben definito perché $|f(x)e^{-2\pi ix\xi}|=|f(x)|$, integrabile per ipotesi). Qui $e^{-2\pi ix\xi}=\cos(2\pi x\xi)-i\sin(2\pi x\xi)$ (formula di Eulero, notebook 9.3), quindi $\hat f$ è a valori complessi in generale, anche per $f$ reale.

**Proprietà elementari.** $\hat f$ è sempre **continua** (dimostrabile con il teorema di convergenza dominata, notebook 13.7: se $\xi_n\to\xi$, l'integranda $f(x)e^{-2\pi ix\xi_n}\to f(x)e^{-2\pi ix\xi}$ puntualmente ed è dominata da $|f(x)|$, integrabile) e **limitata**: $|\hat f(\xi)|\le\int|f(x)|\,dx=\|f\|_1$ per ogni $\xi$.

### La trasformata della gaussiana: un calcolo fondamentale

**Esempio.** Sia $f(x)=e^{-\pi x^2}$ (la gaussiana normalizzata). Si dimostra (usando equazioni differenziali: $\hat f$ soddisfa la stessa ODE di $f$, derivando sotto il segno di integrale e integrando per parti) che $\hat f(\xi)=e^{-\pi\xi^2}$: **la gaussiana è un punto fisso della trasformata di Fourier** (con questa particolare normalizzazione con $2\pi$ nell'esponente). Questo fatto, oltre a essere esteticamente notevole, è lo strumento tecnico per dimostrare molte proprietà generali per densità (approssimando funzioni arbitrarie con gaussiane).

### Il teorema di inversione

**Teorema (inversione di Fourier, enunciato con ipotesi tecniche).** Se $f,\hat f\in L^1(\mathbb R)$, allora $f(x)=\int_{\mathbb R}\hat f(\xi)e^{2\pi ix\xi}\,d\xi$ per quasi ogni $x$ (notebook 13.5) — la stessa formula della trasformata diretta, con il segno dell'esponente invertito.

**Perché servono ipotesi extra.** A differenza della trasformata diretta (definita per ogni $f\in L^1$), l'inversione richiede che **anche** $\hat f$ sia in $L^1$: questa asimmetria è intrinseca alla teoria su $L^1$, e viene risolta in modo più elegante lavorando su $L^2$, come segue.

### Estensione a $L^2$: il teorema di Plancherel

**Teorema (Plancherel).** La trasformata di Fourier, definita inizialmente su $L^1\cap L^2$ (dove entrambe le definizioni hanno senso), si estende in modo unico a un'**isometria** $\mathcal F:L^2(\mathbb R)\to L^2(\mathbb R)$: $\|\hat f\|_2=\|f\|_2$ per ogni $f\in L^2(\mathbb R)$.

**Idea della dimostrazione (schema).** Su $L^1\cap L^2$, si verifica direttamente (con un calcolo che usa il teorema di Fubini, notebook 13.9, per scambiare l'ordine di integrazione) che $\int f\overline g\,dx=\int\hat f\overline{\hat g}\,d\xi$ (identità di Parseval-Plancherel), che con $g=f$ dà $\|f\|_2=\|\hat f\|_2$. Essendo $L^1\cap L^2$ **denso** in $L^2$ (fatto tecnico, dimostrabile troncando le funzioni), e $\mathcal F$ un'isometria su questo sottospazio denso, si estende in modo **unico** a tutto $L^2$ per continuità (un argomento standard di estensione di operatori limitati da un sottospazio denso, collegato allo spirito del teorema di Hahn-Banach, notebook 14.7, anche se qui la tecnica specifica è diversa — l'estensione per densità di un'isometria). $\blacksquare$

**Significato concettuale.** Plancherel dice che $\mathcal F:L^2(\mathbb R)\to L^2(\mathbb R)$ è un **operatore unitario** (isometria suriettiva, con $\mathcal F^{-1}=\mathcal F^*$, l'aggiunto): la trasformata di Fourier è, in un senso preciso, una "rotazione" dello spazio di Hilbert $L^2(\mathbb R)$, esattamente come un cambio di base ortonormale in $\mathbb R^n$ (notebook 6.22) preserva le lunghezze. Questo generalizza l'identità di Parseval discreta del notebook 15.1 al caso continuo.

### La trasformata scambia derivazione e moltiplicazione

**Proprietà (formale, valida su funzioni sufficientemente regolari e decrescenti).** $\widehat{f'}(\xi)=2\pi i\xi\hat f(\xi)$ (derivare in $x$ diventa moltiplicare per $2\pi i\xi$ in frequenza). Questa proprietà — dimostrabile integrando per parti nella definizione, con i termini di bordo che si annullano per $f$ che decade all'infinito — è la ragione **profonda** per cui la trasformata di Fourier è lo strumento standard per risolvere equazioni differenziali (Modulo 8, e i prossimi notebook di questo modulo): trasforma **equazioni differenziali** (che coinvolgono derivate) in **equazioni algebriche** (che coinvolgono solo moltiplicazioni), molto più facili da risolvere.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# Verifica numerica: la gaussiana e' un punto fisso della trasformata di Fourier
def f(x):
    return np.exp(-np.pi * x**2)

def trasformata_numerica(xi, f, limite=10):
    parte_reale = integrate.quad(lambda x: f(x)*np.cos(2*np.pi*x*xi), -limite, limite)[0]
    parte_immaginaria = -integrate.quad(lambda x: f(x)*np.sin(2*np.pi*x*xi), -limite, limite)[0]
    return parte_reale, parte_immaginaria

xi_vals = np.linspace(-2, 2, 20)
trasformata_vals = [trasformata_numerica(xi, f)[0] for xi in xi_vals]  # parte reale (immaginaria = 0 per simmetria)
attesi = np.exp(-np.pi * xi_vals**2)

fig, ax = plt.subplots(figsize=(7,5))
ax.plot(xi_vals, trasformata_vals, 'bo', label='F(f)(xi) numerica')
ax.plot(xi_vals, attesi, 'r-', label='exp(-pi*xi^2) attesa')
ax.legend()
ax.set_title("La trasformata di Fourier della gaussiana e' ancora una gaussiana")
plt.tight_layout()
plt.show()

errore_max = np.max(np.abs(np.array(trasformata_vals) - attesi))
print(f"Errore massimo tra calcolo numerico e formula esatta: {errore_max:.6f}")
```

## Esercizio svolto e commentato

**Problema:** calcola la trasformata di Fourier di $f(x)=e^{-a|x|}$ per $a>0$ fissato, e verifica Plancherel calcolando $\|f\|_2^2$ e confrontandolo con $\int|\hat f(\xi)|^2\,d\xi$.

**Ragionamento:**

**Passo 1 — impostiamo l'integrale.** $\hat f(\xi)=\int_{-\infty}^\infty e^{-a|x|}e^{-2\pi ix\xi}\,dx=\int_{-\infty}^0e^{ax}e^{-2\pi ix\xi}\,dx+\int_0^\infty e^{-ax}e^{-2\pi ix\xi}\,dx$ (spezzando per il valore assoluto).

**Passo 2 — calcoliamo il primo integrale.** $\int_{-\infty}^0e^{(a-2\pi i\xi)x}\,dx=\Big[\dfrac{e^{(a-2\pi i\xi)x}}{a-2\pi i\xi}\Big]_{-\infty}^0=\dfrac1{a-2\pi i\xi}$ (essendo $a>0$, il termine a $-\infty$ si annulla, notebook 5.18 sugli integrali impropri).

**Passo 3 — calcoliamo il secondo integrale.** $\int_0^\infty e^{-(a+2\pi i\xi)x}\,dx=\dfrac1{a+2\pi i\xi}$ (stesso argomento, il termine a $+\infty$ si annulla essendo $a>0$).

**Passo 4 — sommiamo.** $\hat f(\xi)=\dfrac1{a-2\pi i\xi}+\dfrac1{a+2\pi i\xi}=\dfrac{(a+2\pi i\xi)+(a-2\pi i\xi)}{a^2+4\pi^2\xi^2}=\dfrac{2a}{a^2+4\pi^2\xi^2}$ (reale, coerente con il fatto che $f$ è pari e reale — la trasformata di una funzione pari reale è reale).

**Passo 5 — calcoliamo $\|f\|_2^2$.** $\|f\|_2^2=\int_{-\infty}^\infty e^{-2a|x|}\,dx=2\int_0^\infty e^{-2ax}\,dx=2\cdot\dfrac1{2a}=\dfrac1a$.

**Passo 6 — verifichiamo Plancherel calcolando $\int|\hat f|^2\,d\xi$.** $\int_{-\infty}^\infty\Big(\dfrac{2a}{a^2+4\pi^2\xi^2}\Big)^2d\xi=4a^2\int_{-\infty}^\infty\dfrac{d\xi}{(a^2+4\pi^2\xi^2)^2}$. Con la sostituzione $u=2\pi\xi/a$ (quindi $d\xi=a\,du/(2\pi)$): $=4a^2\int_{-\infty}^\infty\dfrac{a\,du/(2\pi)}{a^4(1+u^2)^2}=\dfrac{4}{2\pi a}\int_{-\infty}^\infty\dfrac{du}{(1+u^2)^2}=\dfrac2{\pi a}\cdot\dfrac\pi2=\dfrac1a$ (usando l'integrale standard $\int_{-\infty}^\infty du/(1+u^2)^2=\pi/2$, calcolabile con residui, notebook 9.14, o sostituzione trigonometrica).

**Conclusione:** $\|f\|_2^2=1/a=\int|\hat f(\xi)|^2\,d\xi$: Plancherel verificato esplicitamente su questo esempio.

```python
import numpy as np
from scipy import integrate

a = 2.0

def f(x):
    return np.exp(-a * np.abs(x))

def f_hat(xi):
    return 2*a / (a**2 + 4*np.pi**2*xi**2)

norma_f_L2_quadrato = integrate.quad(lambda x: f(x)**2, -np.inf, np.inf)[0]
norma_fhat_L2_quadrato = integrate.quad(lambda xi: f_hat(xi)**2, -np.inf, np.inf)[0]

print(f"||f||_2^2 (calcolo numerico) = {norma_f_L2_quadrato:.6f}")
print(f"1/a (formula esatta) = {1/a:.6f}")
print(f"\n||f_hat||_2^2 (calcolo numerico) = {norma_fhat_L2_quadrato:.6f}")
print(f"Plancherel verificato: {np.isclose(norma_f_L2_quadrato, norma_fhat_L2_quadrato, rtol=1e-3)}")
```

## Esercizi da risolvere

1. Dimostra che $\hat f$ è reale se $f$ è pari e reale (suggerimento: usa $e^{-2\pi ix\xi}=\cos(2\pi x\xi)-i\sin(2\pi x\xi)$ e la parità per mostrare che la parte immaginaria si annulla per simmetria).
2. Calcola $\hat f$ per $f(x)=\mathbb 1_{[-1/2,1/2]}(x)$ (la funzione indicatrice di un intervallo), ottenendo la funzione **sinc**.
3. Verifica che l'esercizio 2 è coerente con $\hat f$ continua (teoria): controlla che il limite di $\hat f(\xi)$ per $\xi\to0$ esiste ed è finito, anche se la formula ottenuta ha una forma $0/0$ in quel punto.
4. *(più impegnativo)* Usa la proprietà "derivazione ↔ moltiplicazione" della teoria per dimostrare, formalmente, che se $u$ soddisfa l'equazione differenziale $u''-u=0$ su $\mathbb R$ (con $u,u',u''\in L^1\cap L^2$ e decadenti all'infinito), allora $\hat u$ soddisfa $-4\pi^2\xi^2\hat u-\hat u=0$, cioè $\hat u(\xi)(4\pi^2\xi^2+1)=0$ per ogni $\xi$ — e deduci che $\hat u\equiv0$, quindi (per iniettività della trasformata su $L^2$, conseguenza di Plancherel) $u\equiv0$: l'unica soluzione decadente di questa equazione è quella banale.

**1.** $\hat f(\xi)=\int f(x)\cos(2\pi x\xi)\,dx-i\int f(x)\sin(2\pi x\xi)\,dx$. Il secondo integrale: essendo $f$ pari e $\sin(2\pi x\xi)$ dispari (in $x$, per $\xi$ fissato), il prodotto $f(x)\sin(2\pi x\xi)$ è dispari, quindi il suo integrale su $\mathbb R$ (simmetrico) è $0$ (stesso argomento del notebook 15.1 per $b_n=0$ con $f$ pari). Dunque $\hat f(\xi)=\int f(x)\cos(2\pi x\xi)\,dx$, reale.

**2.** $\hat f(\xi)=\int_{-1/2}^{1/2}e^{-2\pi ix\xi}\,dx=\Big[\dfrac{e^{-2\pi ix\xi}}{-2\pi i\xi}\Big]_{-1/2}^{1/2}=\dfrac{e^{-\pi i\xi}-e^{\pi i\xi}}{-2\pi i\xi}=\dfrac{-2i\sin(\pi\xi)}{-2\pi i\xi}=\dfrac{\sin(\pi\xi)}{\pi\xi}=\operatorname{sinc}(\xi)$ (usando $e^{i\theta}-e^{-i\theta}=2i\sin\theta$, formula di Eulero, notebook 9.3).

**3.** Per $\xi\to0$: $\dfrac{\sin(\pi\xi)}{\pi\xi}\to1$ per il limite notevole $\sin(t)/t\to1$ (Modulo 4/5, con $t=\pi\xi\to0$). Dunque $\hat f(0)=1$ (definendo per continuità, o calcolando direttamente $\hat f(0)=\int_{-1/2}^{1/2}1\,dx=1$, coerente): la funzione sinc, pur avendo una forma $0/0$ nella formula chiusa in $\xi=0$, si estende con continuità, coerentemente con il fatto generale che $\hat f$ è sempre continua (teoria).

**4.** Applicando la trasformata a entrambi i membri di $u''-u=0$ (linearità dell'integrale, quindi della trasformata): $\widehat{u''}-\hat u=0$. Usando la proprietà "derivazione↔moltiplicazione" due volte: $\widehat{u'}(\xi)=2\pi i\xi\hat u(\xi)$, quindi $\widehat{u''}(\xi)=2\pi i\xi\cdot\widehat{u'}(\xi)=2\pi i\xi\cdot2\pi i\xi\hat u(\xi)=-4\pi^2\xi^2\hat u(\xi)$. Sostituendo: $-4\pi^2\xi^2\hat u(\xi)-\hat u(\xi)=0$, cioè $\hat u(\xi)(4\pi^2\xi^2+1)=0$ per ogni $\xi$. Essendo $4\pi^2\xi^2+1>0$ sempre (somma di un quadrato non negativo e $1$), deve essere $\hat u(\xi)=0$ per ogni $\xi$, cioè $\hat u\equiv0$. Per Plancherel (teoria), $\|u\|_2=\|\hat u\|_2=0$, quindi $u=0$ quasi ovunque (notebook 13.8): l'unica soluzione (in questo spazio di funzioni decadenti) è quella banale — coerente con il fatto noto dalle equazioni differenziali (Modulo 8) che le soluzioni non banali di $u''-u=0$ sono combinazioni di $e^x,e^{-x}$, nessuna delle quali decade **in entrambe** le direzioni (una cresce quando l'altra decade), quindi nessuna soddisfa le ipotesi di decadimento richieste per stare in $L^1\cap L^2$.

## Riepilogo finale

- La **trasformata di Fourier** $\hat f(\xi)=\int f(x)e^{-2\pi ix\xi}\,dx$ generalizza i coefficienti di Fourier (notebook 15.1) a funzioni non periodiche, sostituendo la somma discreta con un integrale continuo.
- Il **teorema di Plancherel** estende la trasformata a un'isometria su $L^2(\mathbb R)$: $\|\hat f\|_2=\|f\|_2$, l'analogo continuo dell'identità di Parseval.
- La trasformata scambia **derivazione** e **moltiplicazione**: $\widehat{f'}(\xi)=2\pi i\xi\hat f(\xi)$, il fatto che rende la trasformata lo strumento standard per le equazioni differenziali.
- La gaussiana è un punto fisso della trasformata: un fatto usato come mattone tecnico per molte dimostrazioni generali della teoria.
