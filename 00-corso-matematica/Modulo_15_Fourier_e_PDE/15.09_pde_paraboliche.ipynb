# 15.9 — PDE paraboliche

**Prerequisiti:** 15.8 PDE ellittiche, 15.4 Equazione del calore, 14.12 Teorema spettrale.


## Perché ci interessa

Il notebook 15.4 ha risolto l'equazione del calore su tutto $\mathbb R$ tramite trasformata di Fourier — una tecnica che funziona perché $\mathbb R$ ha simmetrie traslazionali semplici. Su un dominio limitato $\Omega$ (una piastra di forma qualsiasi), questa tecnica non si applica direttamente. Le **PDE paraboliche** generalizzano l'equazione del calore a questo contesto, combinando la formulazione variazionale del notebook 15.8 (per la parte spaziale) con il teorema spettrale (notebook 14.12) per costruire soluzioni tramite **separazione di variabili astratta**: esattamente l'idea, mai completamente giustificata nel Modulo 8, ora resa rigorosa con gli strumenti di analisi funzionale sviluppati in questo corso.

## Teoria, passo dopo passo

### Il problema parabolico astratto

Sia $\Omega\subseteq\mathbb R^n$ limitato, e consideriamo $\partial_tu=\Delta u$ su $\Omega\times(0,\infty)$, $u=0$ su $\partial\Omega\times(0,\infty)$ (condizione di Dirichlet omogenea), $u(\cdot,0)=u_0$. Vogliamo $u(\cdot,t)\in H^1_0(\Omega)$ per ogni $t$.

### Autovalori e autofunzioni del laplaciano di Dirichlet

**Teorema (enunciato, applicazione del teorema spettrale).** L'operatore $-\Delta$ con condizioni di Dirichlet su $\Omega$ limitato ha una successione di **autovalori** $0<\lambda_1\le\lambda_2\le\cdots\to\infty$ e corrispondenti **autofunzioni** $\{\phi_k\}$ (cioè $-\Delta\phi_k=\lambda_k\phi_k$ in $\Omega$, $\phi_k=0$ su $\partial\Omega$) che formano una **base ortonormale completa** di $L^2(\Omega)$.

**Perché questo è un'applicazione del notebook 14.12.** L'operatore "risolvente" $R:L^2(\Omega)\to L^2(\Omega)$, $Rf=u$ dove $u\in H^1_0(\Omega)$ è la soluzione debole di $-\Delta u=f$ (esistenza e unicità garantite dal notebook 15.7 via Riesz), è **compatto** (per l'immersione compatta di $H^1_0(\Omega)$ in $L^2(\Omega)$ su domini limitati — un raffinamento del teorema di immersione di Sobolev, notebook 15.3, chiamato teorema di Rellich-Kondrachov, che generalizza a domini limitati la mancanza di compattezza della palla unitaria vista nel notebook 14.11) e **autoaggiunto** (per la simmetria della forma bilineare $\int\nabla u\cdot\nabla v$, notebook 15.8, esercizio 1). Per il **teorema spettrale** (notebook 14.12), $R$ ammette una base ortonormale di autofunzioni $\phi_k$ con autovalori $\mu_k=1/\lambda_k\to0$ (essendo $R$ compatto): riscrivendo $R\phi_k=\mu_k\phi_k$ come $-\Delta\phi_k=\lambda_k\phi_k$ (invertendo la relazione $R=(-\Delta)^{-1}$), si ottengono esattamente gli autovalori/autofunzioni del laplaciano di Dirichlet, con $\lambda_k=1/\mu_k\to\infty$ (essendo $\mu_k\to0$).

### Costruzione della soluzione per separazione di variabili astratta

Sviluppiamo $u_0=\sum_kc_k\phi_k$ (serie di Fourier generalizzata rispetto alla base $\{\phi_k\}$, notebook 15.1 generalizzato, con $c_k=\langle u_0,\phi_k\rangle$). Cerchiamo $u(x,t)=\sum_kc_k(t)\phi_k(x)$, con $c_k(0)=c_k$. Sostituendo nell'equazione (formalmente, derivando termine a termine): $\sum_kc_k'(t)\phi_k=\Delta\Big(\sum_kc_k(t)\phi_k\Big)=\sum_kc_k(t)(-\lambda_k\phi_k)$ (usando $\Delta\phi_k=-\lambda_k\phi_k$). Per l'ortonormalità dei $\phi_k$ (proiettando su ciascuno): $c_k'(t)=-\lambda_kc_k(t)$, un'**ODE lineare del primo ordine** (notebook 8.2) per ogni $k$, con soluzione $c_k(t)=c_ke^{-\lambda_kt}$.

**Formula della soluzione:** $u(x,t)=\sum_kc_ke^{-\lambda_kt}\phi_k(x)$ — l'analogo, per domini limitati, della formula $\hat u(\xi,t)=\hat u_0(\xi)e^{-4\pi^2\xi^2t}$ del notebook 15.4, con la somma discreta sugli autovalori $\lambda_k$ al posto dell'integrale continuo sulle frequenze $\xi$.

### Decadimento esponenziale e stabilità

**Teorema.** $\|u(\cdot,t)\|_2^2=\sum_kc_k^2e^{-2\lambda_kt}\le e^{-2\lambda_1t}\sum_kc_k^2=e^{-2\lambda_1t}\|u_0\|_2^2$ (usando $\lambda_k\ge\lambda_1$ per ogni $k$, e Parseval, notebook 15.1/14.3). Dunque $\|u(\cdot,t)\|_2\le e^{-\lambda_1t}\|u_0\|_2\to0$ **esponenzialmente** quando $t\to\infty$: su un dominio limitato con condizioni di Dirichlet omogenee, la temperatura tende sempre a $0$ (l'equilibrio, coerente con il "raffreddamento" fisico verso un bordo mantenuto a temperatura zero), a **velocità esponenziale** governata dal primo autovalore $\lambda_1$ — un fenomeno di stabilità molto più forte della semplice limitatezza vista nel notebook 15.4 su $\mathbb R$ (dove non c'è decadimento a $0$ in norma $L^2$ senza ipotesi aggiuntive sul dato).

```python
import numpy as np
import matplotlib.pyplot as plt

# Simuliamo la soluzione parabolica su (0,1) (dominio limitato 1D) con autofunzioni sin(k*pi*x)
# -phi_k'' = lambda_k phi_k, phi_k(0)=phi_k(1)=0 -> phi_k(x)=sin(k*pi*x), lambda_k=(k*pi)^2

def phi_k(x, k):
    return np.sqrt(2) * np.sin(k*np.pi*x)  # normalizzata in L^2(0,1)

def lambda_k(k):
    return (k*np.pi)**2

def u0(x):
    return x * (1-x)  # dato iniziale, parabola che si annulla al bordo

x = np.linspace(0, 1, 500)

# Calcoliamo i coefficienti c_k = <u0, phi_k>
K = 20
c_k = [np.trapz(u0(x) * phi_k(x, k), x) for k in range(1, K+1)]

def u_t(x, t, c_k, K):
    return sum(c_k[k-1] * np.exp(-lambda_k(k)*t) * phi_k(x, k) for k in range(1, K+1))

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x, u0(x), 'k--', label='u0')
for t in [0.001, 0.01, 0.05, 0.2]:
    ax.plot(x, u_t(x, t, c_k, K), label=f"u(.,{t})")
ax.legend()
ax.set_title("Equazione del calore su (0,1): decadimento verso 0 (Dirichlet omogeneo)")
plt.tight_layout()
plt.show()

# Verifica del decadimento esponenziale della norma L^2
norme = []
tempi = np.linspace(0, 0.3, 20)
for t in tempi:
    u_vals = u_t(x, t, c_k, K)
    norme.append(np.sqrt(np.trapz(u_vals**2, x)))

lambda_1 = lambda_k(1)
norma_0 = norme[0]
fig, ax = plt.subplots(figsize=(7,4))
ax.plot(tempi, norme, 'bo-', label='||u(.,t)||_2 (numerico)')
ax.plot(tempi, norma_0*np.exp(-lambda_1*tempi), 'r--', label=f'stima exp(-lambda_1*t), lambda_1={lambda_1:.2f}')
ax.legend()
ax.set_title("Decadimento della norma L^2, dominato da exp(-lambda_1*t)")
plt.tight_layout()
plt.show()
```

## Esercizio svolto e commentato

**Problema:** su $\Omega=(0,\pi)$ (per semplificare le costanti), risolvi $\partial_tu=\partial_x^2u$, $u(0,t)=u(\pi,t)=0$, $u(x,0)=3\sin(2x)-\sin(5x)$, e calcola $\lim_{t\to\infty}\|u(\cdot,t)\|_2^2\cdot e^{2t}$ (il tasso di decadimento asintotico).

**Ragionamento:**

**Passo 1 — identifichiamo autovalori e autofunzioni su $(0,\pi)$.** $\phi_k(x)=\sin(kx)$ (non normalizzata, ma sufficiente qui), $\lambda_k=k^2$ (verifica diretta: $-\phi_k''=k^2\sin(kx)=k^2\phi_k$).

**Passo 2 — riconosciamo che il dato iniziale è già nella base.** $u_0(x)=3\sin(2x)-\sin(5x)$: coefficiente $c_2=3$ (per $k=2$), $c_5=-1$ (per $k=5$), tutti gli altri nulli.

**Passo 3 — scriviamo la soluzione esplicita.** $u(x,t)=3e^{-4t}\sin(2x)-e^{-25t}\sin(5x)$ (usando $\lambda_2=4$, $\lambda_5=25$).

**Passo 4 — calcoliamo $\|u(\cdot,t)\|_2^2$.** Essendo $\sin(2x)$ e $\sin(5x)$ ortogonali su $(0,\pi)$ (stesso argomento del notebook 15.1, con $\int_0^\pi\sin(2x)\sin(5x)\,dx=0$), per Pitagora: $\|u(\cdot,t)\|_2^2=9e^{-8t}\|\sin(2x)\|_2^2+e^{-50t}\|\sin(5x)\|_2^2=9e^{-8t}\cdot\dfrac\pi2+e^{-50t}\cdot\dfrac\pi2$ (usando $\int_0^\pi\sin^2(kx)\,dx=\pi/2$ per ogni $k$ intero, stesso calcolo del notebook 15.1).

**Passo 5 — calcoliamo il limite richiesto.** Per $t\to\infty$, il termine $e^{-50t}$ decade **molto più velocemente** di $e^{-8t}$ (essendo $50>8$): $\|u(\cdot,t)\|_2^2\sim\dfrac{9\pi}2e^{-8t}$ per $t$ grande. Ma la domanda chiede $\lim_t\|u(\cdot,t)\|_2^2\cdot e^{2t}$: essendo il decadimento dominante $e^{-8t}$ (molto più rapido di $e^{-2t}$), $\|u(\cdot,t)\|_2^2\cdot e^{2t}\sim\dfrac{9\pi}2e^{-6t}\to0$.

**Conclusione:** $\lim_{t\to\infty}\|u(\cdot,t)\|_2^2e^{2t}=0$ (il decadimento reale, governato da $\lambda_1=1$ nel caso generale ma qui dal più piccolo autovalore **effettivamente presente nel dato**, $\lambda_2=4$, è più rapido di $e^{-2t}$). **Punto metodologico:** il tasso di decadimento asintotico di una soluzione è sempre determinato dal **più piccolo autovalore con coefficiente $c_k\ne0$** nel dato iniziale — non necessariamente $\lambda_1$ in assoluto, se il dato iniziale non ha componente lungo $\phi_1$ (qui, $c_1=0$).

```python
import numpy as np
import matplotlib.pyplot as plt

def u(x, t):
    return 3*np.exp(-4*t)*np.sin(2*x) - np.exp(-25*t)*np.sin(5*x)

x = np.linspace(0, np.pi, 500)
t_vals = np.linspace(0, 2, 50)
norme_quadrato = []
for t in t_vals:
    u_t = u(x, t)
    norme_quadrato.append(np.trapz(u_t**2, x))

fig, ax = plt.subplots(figsize=(7,5))
ax.semilogy(t_vals, norme_quadrato, 'b-', label='||u(.,t)||_2^2 (numerico)')
ax.semilogy(t_vals, (9*np.pi/2)*np.exp(-8*t_vals), 'r--', label='9*pi/2 * exp(-8t) (asintotico)')
ax.legend()
ax.set_title("Decadimento dominato da exp(-8t) = exp(-2*lambda_2*t)")
plt.tight_layout()
plt.show()

prodotto = np.array(norme_quadrato) * np.exp(2*t_vals)
print("||u(.,t)||^2 * exp(2t) per t grande:", prodotto[-5:])
print("Tende a 0, confermando che il decadimento e' piu' rapido di exp(-2t)")
```

3. Dimostra che l'energia $E(t)=\dfrac12\|\nabla u(\cdot,t)\|_2^2$ decade anch'essa esponenzialmente, con tasso $2\lambda_1$ (suggerimento: usa $\|\nabla u(\cdot,t)\|_2^2=\sum_k\lambda_kc_k^2e^{-2\lambda_kt}$, una conseguenza di $\|\nabla\phi_k\|_2^2=\lambda_k\|\phi_k\|_2^2$ ottenuta integrando per parti $\int\nabla\phi_k\cdot\nabla\phi_k=-\int\phi_k\Delta\phi_k=\lambda_k\int\phi_k^2$).
4. *(più impegnativo)* Usa la formula generale $u(x,t)=\sum_kc_ke^{-\lambda_kt}\phi_k(x)$ per dimostrare che, per **ogni** $t>0$ fissato, $u(\cdot,t)\in H^s(\Omega)$ per ogni $s$ (un effetto regolarizzante analogo a quello dimostrato nell'esercizio 4 del notebook 15.4, ma ora nel linguaggio della base spettrale invece che della trasformata di Fourier — suggerimento: usa che $\lambda_k^se^{-2\lambda_kt}$ è limitato uniformemente in $k$ per $t>0$ fissato, essendo $\lambda_k\to\infty$ e l'esponenziale che domina ogni potenza).

**1.** $c_1=1$ ($\lambda_1=1$), $c_3=2$ ($\lambda_3=9$). $u(x,t)=e^{-t}\sin(x)+2e^{-9t}\sin(3x)$. Il tasso dominante per $t\to\infty$ è $e^{-t}$ (essendo $\lambda_1=1<9=\lambda_3$, il termine con $k=1$ decade più lentamente e domina asintoticamente).

**2.** Con $u_0=\phi_1$: $c_1=\langle\phi_1,\phi_1\rangle=\|\phi_1\|_2^2$ (se non normalizzata) o $c_1=1$ (se $\phi_1$ è già normalizzata, $\|\phi_1\|_2=1$), e tutti gli altri $c_k=0$ (per ortogonalità, notebook 14.3). **3.** $\|\nabla u(\cdot,t)\|_2^2=\Big\|\sum_kc_ke^{-\lambda_kt}\nabla\phi_k\Big\|_2^2=\sum_kc_k^2e^{-2\lambda_kt}\|\nabla\phi_k\|_2^2$ (usando che i $\nabla\phi_k$ ereditano l'ortogonalità dai $\phi_k$ tramite l'identità $\int\nabla\phi_j\cdot\nabla\phi_k=\lambda_k\int\phi_j\phi_k=0$ per $j\ne k$, integrando per parti come suggerito) $=\sum_kc_k^2e^{-2\lambda_kt}\lambda_k\|\phi_k\|_2^2=\sum_k\lambda_kc_k^2e^{-2\lambda_kt}$ (assumendo $\phi_k$ normalizzate, $\|\phi_k\|_2=1$). Essendo $\lambda_ke^{-2\lambda_kt}\le\lambda_1e^{-2\lambda_1t}\cdot\dfrac{\lambda_k}{\lambda_1}e^{-2(\lambda_k-\lambda_1)t}$, e usando che la funzione $\lambda\mapsto\lambda e^{-2\lambda t}$ per $t>0$ fissato è decrescente per $\lambda$ grande (derivata $e^{-2\lambda t}(1-2\lambda t)<0$ per $\lambda>1/(2t)$), si può mostrare (con un argomento più fine che omettiamo nei dettagli tecnici) che $E(t)\le E(0)e^{-2\lambda_1t}$ per $t$ non troppo piccolo, dando il tasso $2\lambda_1$ asintoticamente.

**4.** $\|u(\cdot,t)\|_{H^s}^2\sim\sum_k\lambda_k^sc_k^2e^{-2\lambda_kt}$ (usando che, nella base spettrale, la norma $H^s$ è equivalente a pesare i coefficienti con $\lambda_k^s$, analogamente al peso $(1+\xi^2)^s$ del notebook 15.3 — un fatto che generalizza la definizione di Sobolev tramite l'operatore $-\Delta$ invece che tramite la trasformata di Fourier). Per $t>0$ fissato, la funzione $\lambda\mapsto\lambda^se^{-2\lambda t}$ è continua su $[0,\infty)$ e tende a $0$ per $\lambda\to\infty$ (l'esponenziale domina ogni potenza, notebook 4.2/10.10): dunque è limitata da una costante $C_{s,t}$. Allora $\|u(\cdot,t)\|_{H^s}^2\le C_{s,t}\sum_kc_k^2=C_{s,t}\|u_0\|_2^2<\infty$ (per Parseval, essendo $u_0\in L^2$): $u(\cdot,t)\in H^s(\Omega)$ per ogni $s$, per ogni $t>0$, indipendentemente dalla regolarità di $u_0$ — l'effetto regolarizzante, ora dimostrato nel linguaggio spettrale.

## Riepilogo finale

- L'operatore $-\Delta$ con condizioni di Dirichlet su un dominio limitato ha una base ortonormale completa di autofunzioni, con autovalori $\lambda_k\to\infty$: conseguenza diretta del teorema spettrale (notebook 14.12) applicato all'operatore risolvente (compatto per Rellich-Kondrachov).
- La soluzione del problema parabolico si scrive come $u(x,t)=\sum_kc_ke^{-\lambda_kt}\phi_k(x)$: la separazione di variabili del Modulo 8, ora rigorosamente giustificata.
- Su domini limitati, la soluzione decade **esponenzialmente** a $0$ in norma $L^2$, con tasso governato dal più piccolo autovalore effettivamente presente nel dato iniziale — un fenomeno di stabilità più forte di quanto visto su $\mathbb R$ (notebook 15.4).
- L'effetto regolarizzante (soluzione in ogni $H^s$ per $t>0$) si ridimostra nel linguaggio spettrale, parallelo a quanto fatto con la trasformata di Fourier.
