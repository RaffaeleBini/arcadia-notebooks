**Prerequisiti:** 15.2 Trasformata di Fourier (rigorosa), 14.3 Spazi di Hilbert, 10.14 Derivazione rigorosa.


## Perché ci interessa

Per risolvere un'equazione differenziale come $u''-u=f$ (notebook 15.2, esercizio 4), serve spesso lavorare con funzioni che hanno derivate "abbastanza buone" — ma non necessariamente derivabili in senso classico (notebook 10.14). Gli **spazi di Sobolev** $H^s$ risolvono questo problema: usano la trasformata di Fourier per **definire** la regolarità in modo quantitativo e continuo (anche per $s$ non intero), senza richiedere che le derivate esistano puntualmente. Questo notebook costruisce $H^s(\mathbb R)$ come spazio di Hilbert (notebook 14.3), preparando il linguaggio necessario per formulare rigorosamente le equazioni alle derivate parziali dei prossimi notebook.

## Teoria, passo dopo passo

### L'idea: misurare la regolarità in frequenza

Per il notebook 15.2, $\widehat{f'}(\xi)=2\pi i\xi\hat f(\xi)$: derivare corrisponde a moltiplicare per $\xi$ in frequenza. Dunque, se $\hat f(\xi)$ decade rapidamente per $|\xi|\to\infty$, anche $\xi\hat f(\xi)$ (la trasformata di $f'$) decade abbastanza da essere ancora in $L^2$: **il decadimento di $\hat f$ controlla quante derivate di $f$ stanno in $L^2$**. Questa osservazione, resa quantitativa, è la definizione degli spazi di Sobolev.

### Definizione dello spazio di Sobolev $H^s(\mathbb R)$

Per $s\ge0$ (anche non intero), $H^s(\mathbb R)=\Big\{f\in L^2(\mathbb R):\int_{\mathbb R}(1+\xi^2)^s|\hat f(\xi)|^2\,d\xi<\infty\Big\}$, con prodotto scalare $\langle f,g\rangle_{H^s}=\int_{\mathbb R}(1+\xi^2)^s\hat f(\xi)\overline{\hat g(\xi)}\,d\xi$ e norma $\|f\|_{H^s}=\Big(\int(1+\xi^2)^s|\hat f(\xi)|^2\,d\xi\Big)^{1/2}$.

**Coerenza con $L^2$.** $H^0(\mathbb R)=L^2(\mathbb R)$ (con $s=0$, il fattore $(1+\xi^2)^0=1$ non pesa nulla, e la norma $H^0$ coincide con $\|\hat f\|_2=\|f\|_2$ per Plancherel, notebook 15.2).

### $H^s$ è uno spazio di Hilbert

**Teorema.** $(H^s(\mathbb R),\langle\cdot,\cdot\rangle_{H^s})$ è uno spazio di Hilbert.

**Dimostrazione (schema).** Gli assiomi di prodotto scalare (notebook 14.3) si verificano direttamente dalla definizione (positività: l'integranda $(1+\xi^2)^s|\hat f|^2\ge0$; simmetria e bilinearità ereditate da quelle di $L^2$). Per la completezza: se $(f_n)$ è di Cauchy in $H^s$, allora $((1+\xi^2)^{s/2}\hat f_n)_n$ è di Cauchy in $L^2$ (per definizione della norma $H^s$); per completezza di $L^2$ (Riesz-Fischer, notebook 13.8), converge a una funzione $g\in L^2$; ponendo $\hat f=(1+\xi^2)^{-s/2}g$ e usando l'iniettività della trasformata di Fourier su $L^2$ (conseguenza di Plancherel, notebook 15.2), si recupera $f\in H^s$ con $f_n\to f$ in norma $H^s$. $\blacksquare$

### $H^s$ per $s$ intero: derivate deboli in $L^2$

**Teorema (per $s=k$ intero non negativo).** $f\in H^k(\mathbb R)$ se e solo se $f,f',f'',\ldots,f^{(k)}\in L^2(\mathbb R)$ (nel senso delle **derivate deboli**, che coincidono con quelle classiche se $f$ è sufficientemente regolare, ma hanno senso anche per funzioni meno regolari — un concetto che generalizza la derivata classica in modo analogo a come l'integrale di Lebesgue, notebook 13.4, generalizza quello di Riemann).

**Idea della dimostrazione.** $\|f\|_{H^k}^2=\int(1+\xi^2)^k|\hat f(\xi)|^2\,d\xi$; espandendo $(1+\xi^2)^k=\sum_{j=0}^k\binom kj\xi^{2j}$ (binomio di Newton, Modulo 1), questo diventa una combinazione di $\int\xi^{2j}|\hat f(\xi)|^2\,d\xi=\|\widehat{f^{(j)}}\|_2^2=\|f^{(j)}\|_2^2$ (usando la proprietà derivazione↔moltiplicazione e Plancherel, notebook 15.2, per ogni $j\le k$): dunque $\|f\|_{H^k}^2$ è (a meno di costanti binomiali) equivalente a $\sum_{j=0}^k\|f^{(j)}\|_2^2$, finito se e solo se ogni derivata fino all'ordine $k$ sta in $L^2$.

### Il teorema di immersione di Sobolev (enunciato)

**Teorema (immersione di Sobolev, caso $\mathbb R$, enunciato).** Se $s>1/2$, ogni $f\in H^s(\mathbb R)$ ha un rappresentante **continuo** (e limitato). Più precisamente, $H^s(\mathbb R)\subseteq C(\mathbb R)\cap L^\infty(\mathbb R)$ per $s>1/2$.

**Perché la soglia $1/2$.** Intuitivamente: $f(x)=\int\hat f(\xi)e^{2\pi ix\xi}\,d\xi$ (inversione, notebook 15.2) richiede $\hat f\in L^1$ per essere ben definita puntualmente. Per Cauchy-Schwarz (notebook 6.20/14.3): $\int|\hat f(\xi)|\,d\xi=\int(1+\xi^2)^{s/2}|\hat f(\xi)|\cdot(1+\xi^2)^{-s/2}\,d\xi\le\|f\|_{H^s}\cdot\Big(\int(1+\xi^2)^{-s}\,d\xi\Big)^{1/2}$, e l'ultimo integrale converge se e solo se $2s>1$ (criterio di convergenza per integrali impropri del tipo $\int|\xi|^{-2s}d\xi$ all'infinito, Modulo 5/13.8), cioè $s>1/2$: **esattamente** la soglia del teorema. Questo esempio mostra come la teoria di Sobolev traduca una domanda qualitativa ("$f$ è continua?") in un preciso calcolo quantitativo sul decadimento di $\hat f$.

```python
import numpy as np
from scipy import integrate

# Verifica numerica della soglia s=1/2 nell'immersione di Sobolev, tramite l'integrale di convergenza
def integrale_convergenza(s, limite=1e6):
    # integrale di (1+xi^2)^(-s) su tutto R, calcolato numericamente (raddoppiando la meta' positiva)
    val, _ = integrate.quad(lambda xi: (1+xi**2)**(-s), 0, limite, limit=200)
    return 2 * val

for s in [0.3, 0.49, 0.5, 0.51, 0.7, 1.0]:
    try:
        val = integrale_convergenza(s, limite=1e5)
        print(f"s={s}: integrale di (1+xi^2)^(-s) approssimato = {val:.4f}")
    except Exception as e:
        print(f"s={s}: errore ({e})")

print("\nPer s<=0.5 l'integrale cresce indefinitamente al crescere del limite (diverge);")
print("per s>0.5 converge a un valore finito: questa e' la soglia del teorema di immersione")

# Confronto diretto: al crescere del limite di integrazione, s=0.4 diverge, s=0.6 converge
for s in [0.4, 0.6]:
    for lim in [1e2, 1e4, 1e6]:
        val = integrale_convergenza(s, limite=lim)
        print(f"s={s}, limite={lim:.0e}: integrale = {val:.4f}")
```

## Esercizio svolto e commentato

**Problema:** verifica che $f(x)=e^{-\pi x^2}$ (la gaussiana, notebook 15.2) appartiene a $H^s(\mathbb R)$ per **ogni** $s\ge0$, cioè è "infinitamente Sobolev regolare".

**Ragionamento:**

**Passo 1 — usiamo il calcolo esplicito del notebook 15.2.** $\hat f(\xi)=e^{-\pi\xi^2}$ (la gaussiana è un punto fisso della trasformata).

**Passo 2 — calcoliamo $\|f\|_{H^s}^2$.** $\|f\|_{H^s}^2=\int_{\mathbb R}(1+\xi^2)^se^{-2\pi\xi^2}\,d\xi$ (usando $|\hat f(\xi)|^2=(e^{-\pi\xi^2})^2=e^{-2\pi\xi^2}$, essendo $\hat f$ reale).

**Passo 3 — verifichiamo la convergenza per ogni $s\ge0$.** Il fattore $(1+\xi^2)^s$ cresce **polinomialmente** in $|\xi|$ (per qualunque $s$ fissato), mentre $e^{-2\pi\xi^2}$ decade **esponenzialmente**: per un fatto standard sui limiti (l'esponenziale domina ogni potenza, notebook 4.2/10.10, generalizzato a potenze reali $s$), il prodotto $(1+\xi^2)^se^{-2\pi\xi^2}\to0$ rapidamente per $|\xi|\to\infty$, e l'integrale su tutto $\mathbb R$ converge (essendo l'integranda continua e a decadimento più che sufficiente per l'integrabilità, un confronto diretto con $e^{-\pi\xi^2}$ stesso, che è integrabile).

**Conclusione:** $\|f\|_{H^s}<\infty$ per ogni $s\ge0$: la gaussiana appartiene a $H^s(\mathbb R)$ per ogni $s$, cioè è infinitamente derivabile in senso $L^2$ (coerente con il fatto, noto dal Modulo 5, che la gaussiana è di classe $C^\infty$ in senso classico, con tutte le derivate ancora prodotto di un polinomio per una gaussiana, quindi sempre in $L^2$). **Portata:** questo esempio mostra che le funzioni "lisce e a decadimento rapido" (come la gaussiana, l'esempio guida della cosiddetta classe di Schwartz) sono automaticamente in ogni $H^s$: la teoria di Sobolev diventa davvero interessante e informativa per funzioni **meno regolari**, come vedremo nei prossimi notebook con soluzioni di equazioni alle derivate parziali che potrebbero non essere lisce ovunque.

```python
import numpy as np
from scipy import integrate

def norma_Hs_quadrato(s, limite=20):
    integranda = lambda xi: (1+xi**2)**s * np.exp(-2*np.pi*xi**2)
    val, _ = integrate.quad(integranda, -limite, limite)
    return val

for s in [0, 1, 2, 5, 10, 50]:
    val = norma_Hs_quadrato(s)
    print(f"s={s}: ||f||_H^s^2 = {val:.6f} (finita per ogni s)")
```

**1.** Se $f\in H^{s_1}$, allora $\int(1+\xi^2)^{s_1}|\hat f(\xi)|^2\,d\xi<\infty$. Essendo $1+\xi^2\ge1$ per ogni $\xi$, e $s_1\ge s_2\ge0$: $(1+\xi^2)^{s_2}\le(1+\xi^2)^{s_1}$ (una base $\ge1$ elevata a un esponente minore dà un risultato minore o uguale, proprietà delle potenze, Modulo 0/1). Dunque $(1+\xi^2)^{s_2}|\hat f(\xi)|^2\le(1+\xi^2)^{s_1}|\hat f(\xi)|^2$ per ogni $\xi$, quindi $\int(1+\xi^2)^{s_2}|\hat f|^2\,d\xi\le\int(1+\xi^2)^{s_1}|\hat f|^2\,d\xi<\infty$ (monotonia dell'integrale, notebook 13.4): $f\in H^{s_2}$.

**2.** Dall'esercizio 2 del notebook 15.2 (con riscalamento per l'intervallo $[-1,1]$ invece di $[-1/2,1/2]$): $\hat f(\xi)=\dfrac{\sin(2\pi\xi)}{\pi\xi}$, che per $|\xi|$ grande si comporta come $O(1/|\xi|)$ (essendo $|\sin|\le1$). Dunque $(1+\xi^2)^s|\hat f(\xi)|^2\sim\xi^{2s}\cdot\dfrac1{\xi^2}=\xi^{2s-2}$ per $|\xi|$ grande. L'integrale $\int^\infty\xi^{2s-2}\,d\xi$ converge se e solo se $2s-2<-1$, cioè $s<1/2$ (criterio standard di convergenza degli integrali impropri con potenze, Modulo 5/13.8). Dunque per $s\ge1/2$ (in particolare $s>1/2$), l'integrale diverge: $f\notin H^s(\mathbb R)$ per $s>1/2$, coerentemente con il teorema di immersione (essendo $f$ discontinua, non può avere un rappresentante continuo, quindi non può stare in $H^s$ con $s>1/2$).

**3.** Per $s=1$: $\|f\|_{H^1}^2=\int(1+\xi^2)|\hat f(\xi)|^2\,d\xi=\int|\hat f(\xi)|^2\,d\xi+\int\xi^2|\hat f(\xi)|^2\,d\xi=\|f\|_2^2+\|\widehat{f'}\|_2^2/(2\pi)^2\cdot(2\pi)^2$ — più precisamente, usando $\widehat{f'}(\xi)=2\pi i\xi\hat f(\xi)$, si ha $|\widehat{f'}(\xi)|^2=4\pi^2\xi^2|\hat f(\xi)|^2$, quindi $\int\xi^2|\hat f|^2\,d\xi=\dfrac1{4\pi^2}\|\widehat{f'}\|_2^2=\dfrac1{4\pi^2}\|f'\|_2^2$ (Plancherel, notebook 15.2). Dunque $\|f\|_{H^1}^2=\|f\|_2^2+\dfrac1{4\pi^2}\|f'\|_2^2$, finito se e solo se sia $\|f\|_2$ sia $\|f'\|_2$ sono finiti: $f\in H^1\iff f,f'\in L^2$.

**4.** Applicando la trasformata di Fourier all'equazione (come nell'esercizio 4 del notebook 15.2): $4\pi^2\xi^2\hat u(\xi)+\hat u(\xi)=\hat f(\xi)$, cioè $\hat u(\xi)=\dfrac{\hat f(\xi)}{4\pi^2\xi^2+1}$. Allora $\|u\|_{H^2}^2=\int(1+\xi^2)^2|\hat u(\xi)|^2\,d\xi=\int(1+\xi^2)^2\dfrac{|\hat f(\xi)|^2}{(4\pi^2\xi^2+1)^2}\,d\xi$. Essendo $(1+\xi^2)^2$ e $(4\pi^2\xi^2+1)^2$ entrambi polinomi di grado $4$ in $\xi$ con termine dominante positivo, il loro rapporto è **limitato** per ogni $\xi$ (un fatto verificabile con un confronto diretto dei coefficienti, o osservando che il rapporto tende a una costante finita per $\xi\to\infty$ e è continuo, quindi limitato su ogni compatto): esiste $C$ con $(1+\xi^2)^2/(4\pi^2\xi^2+1)^2\le C$ per ogni $\xi$. Dunque $\|u\|_{H^2}^2\le C\int|\hat f(\xi)|^2\,d\xi=C\|f\|_2^2<\infty$ (essendo $f\in L^2$ per ipotesi): $u\in H^2(\mathbb R)$. Essendo $2>1/2$, per il teorema di immersione di Sobolev, $u$ ha un rappresentante continuo.

## Riepilogo finale

- Gli **spazi di Sobolev** $H^s(\mathbb R)$ misurano la regolarità di una funzione tramite il decadimento della sua trasformata di Fourier, generalizzando la nozione di "derivate in $L^2$" a esponenti $s$ anche non interi.
- $H^s$ è sempre uno spazio di **Hilbert**; per $s=k$ intero, $H^k$ coincide con le funzioni che hanno derivate (deboli) fino all'ordine $k$ in $L^2$.
- Il **teorema di immersione di Sobolev** collega la regolarità "in frequenza" ($s>1/2$ su $\mathbb R$) alla regolarità classica (continuità): un ponte quantitativo preciso tra le due nozioni.
- Funzioni lisce a decadimento rapido (come la gaussiana) stanno in **ogni** $H^s$; funzioni con discontinuità (come un'indicatrice) escono da $H^s$ appena $s$ supera la soglia critica $1/2$.
