## Perché ci interessa

Nel notebook 8.9 abbiamo introdotto informalmente le serie di Fourier: scrivere una funzione periodica come somma di seni e coseni. Nel notebook 14.12 abbiamo chiuso il Modulo 14 promettendo che le serie di Fourier sono "precisamente la decomposizione spettrale dell'operatore di derivazione seconda su funzioni periodiche". Ora possiamo mantenere questa promessa: lo spazio $L^2$ delle funzioni periodiche (notebook 13.8) è uno spazio di Hilbert (notebook 14.3), il sistema trigonometrico $\{1,\cos(nx),\sin(nx)\}$ è una base ortonormale (dopo normalizzazione) per questo spazio, e la serie di Fourier di una funzione è semplicemente la sua espansione rispetto a questa base — un'istanza diretta della teoria generale degli spazi di Hilbert, non più un trucco calcolistico ad hoc.

## Teoria, passo dopo passo

### Lo spazio $L^2(\mathbb T)$ e il sistema trigonometrico

Sia $\mathbb T=[-\pi,\pi]$ (con gli estremi identificati, notebook 12.10 — la circonferenza) e $L^2(\mathbb T)$ lo spazio di Hilbert (notebook 14.3, con $\langle f,g\rangle=\int_{-\pi}^\pi fg\,dx$) delle funzioni a quadrato integrabile e periodiche di periodo $2\pi$. Consideriamo il sistema $\{e_0,c_n,s_n\}_{n\ge1}$ con $e_0(x)=\dfrac1{\sqrt{2\pi}}$, $c_n(x)=\dfrac{\cos(nx)}{\sqrt\pi}$, $s_n(x)=\dfrac{\sin(nx)}{\sqrt\pi}$.

**Teorema (ortonormalità).** Questo sistema è ortonormale in $L^2(\mathbb T)$.

**Dimostrazione (calcolo diretto, formule di prostaferesi).** $\langle c_n,c_m\rangle=\dfrac1\pi\int_{-\pi}^\pi\cos(nx)\cos(mx)\,dx$. Usando $\cos(nx)\cos(mx)=\dfrac12[\cos((n-m)x)+\cos((n+m)x)]$ (formula di prostaferesi, Modulo 2): per $n\ne m$, entrambi gli integrali $\int_{-\pi}^\pi\cos(kx)\,dx=0$ per $k\ne0$ intero (l'integrale di un periodo completo di un coseno non costante si annulla); per $n=m$, $\int_{-\pi}^\pi\cos(0)\,dx+\int_{-\pi}^\pi\cos(2nx)\,dx=2\pi+0=2\pi$, quindi $\langle c_n,c_n\rangle=\dfrac1\pi\cdot\dfrac{2\pi}2=1$ (normalizzato correttamente). Calcoli analoghi per $\langle s_n,s_m\rangle$ e $\langle e_0,e_0\rangle$, e $\langle c_n,s_m\rangle=0$ sempre (integrale di una funzione dispari su un intervallo simmetrico, essendo $\cos(nx)\sin(mx)$ dispari). $\blacksquare$

### I coefficienti di Fourier come proiezioni ortogonali

Per il teorema della proiezione ortogonale (notebook 14.3), data $f\in L^2(\mathbb T)$, la sua proiezione sul sottospazio generato da $\{e_0,c_1,s_1,\ldots,c_N,s_N\}$ è $\sum_{n=0}^N\langle f,\cdot\rangle\cdot(\text{base})$: esplicitamente, definendo $a_0=\langle f,e_0\rangle/\sqrt{2\pi}\cdot2$ ecc. (con le normalizzazioni classiche), si riottengono le formule familiari del notebook 8.9: $a_n=\dfrac1\pi\int_{-\pi}^\pi f(x)\cos(nx)\,dx$, $b_n=\dfrac1\pi\int_{-\pi}^\pi f(x)\sin(nx)\,dx$, e la serie di Fourier è $\dfrac{a_0}2+\sum_{n=1}^\infty(a_n\cos(nx)+b_n\sin(nx))$.

### Il sistema trigonometrico è una base ortonormale completa

**Teorema (enunciato, dimostrazione tramite Stone-Weierstrass, che esula da questo notebook).** Il sistema trigonometrico è **completo** in $L^2(\mathbb T)$: non esiste $f\ne0$ in $L^2(\mathbb T)$ ortogonale a **tutti** gli elementi del sistema. Questo garantisce, per la teoria generale delle basi ortonormali (collegata al notebook 6.22 e al teorema spettrale del notebook 14.12), che la serie di Fourier di $f$ **converge a $f$ in norma $L^2$** per ogni $f\in L^2(\mathbb T)$ — non puntualmente in generale, ma nel senso della norma dell'integrale (notebook 13.8).

### L'identità di Parseval

**Teorema (Parseval).** $\|f\|_2^2=\dfrac{a_0^2}2\pi+\pi\sum_{n=1}^\infty(a_n^2+b_n^2)$ (con le normalizzazioni classiche), o equivalentemente $\|f\|_2^2=\sum_k|\langle f,e_k\rangle|^2$ nella notazione ortonormale astratta.

**Dimostrazione.** Conseguenza diretta della completezza della base (teorema sopra) e del teorema di Pitagora generalizzato (notebook 14.3): se $f=\sum_k\langle f,e_k\rangle e_k$ (convergenza $L^2$), allora $\|f\|_2^2=\sum_k\langle f,e_k\rangle^2$ (Pitagora, essendo gli $e_k$ ortonormali). $\blacksquare$ Questa identità generalizza, allo spazio $L^2(\mathbb T)$ concreto, la stessa identità di Pitagora usata nel notebook 14.3 per lo spazio astratto $\ell^2$.

### Convergenza puntuale: il ruolo della regolarità

La convergenza $L^2$ (garantita sempre) è più debole della convergenza puntuale. **Teorema (Dirichlet, enunciato).** Se $f$ è di classe $C^1$ a tratti, la serie di Fourier converge puntualmente a $f(x)$ in ogni punto di continuità, e alla media dei limiti destro/sinistro nei punti di salto (il fenomeno di Gibbs, notebook 10.17, descrive l'oscillazione che persiste vicino ai salti anche al crescere di $N$, senza scomparire in ampiezza — solo restringendosi in larghezza).

```python
import numpy as np
import matplotlib.pyplot as plt

# Serie di Fourier dell'onda quadra: f(x) = 1 su (0,pi), -1 su (-pi,0)
def onda_quadra(x):
    return np.sign(np.sin(x))

def somma_parziale_fourier(x, N):
    # per l'onda quadra, a_n=0 (funzione dispari), b_n = 4/(n*pi) per n dispari, 0 per n pari
    s = np.zeros_like(x)
    for n in range(1, N+1, 2):  # solo n dispari
        s += (4/(n*np.pi)) * np.sin(n*x)
    return s

x = np.linspace(-np.pi, np.pi, 1000)
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x, onda_quadra(x), 'k--', label='f(x) onda quadra', linewidth=1)
for N in [1, 5, 21, 101]:
    ax.plot(x, somma_parziale_fourier(x, N), label=f"S_{N}(x)")
ax.legend()
ax.set_title("Convergenza della serie di Fourier dell'onda quadra (fenomeno di Gibbs vicino ai salti)")
plt.tight_layout()
plt.show()

# Verifica numerica di Parseval: ||f||_2^2 = pi * sum(b_n^2) per l'onda quadra
norma_f_quadrato = np.trapz(onda_quadra(x)**2, x)
somma_bn_quadro = sum((4/(n*np.pi))**2 for n in range(1, 10000, 2))
print(f"||f||_2^2 (integrale diretto) = {norma_f_quadrato:.4f}")
print(f"pi * sum(b_n^2) (Parseval, troncata a 10000 termini) = {np.pi * somma_bn_quadro:.4f}")
```

## Esercizio svolto e commentato

**Problema:** calcola la serie di Fourier di $f(x)=x$ su $(-\pi,\pi)$ (estesa periodicamente), e usa Parseval per calcolare $\sum_{n=1}^\infty1/n^2$.

**Ragionamento:**

**Passo 1 — osserviamo che $f$ è dispari.** $f(-x)=-x=-f(x)$: essendo $f$ dispari e $\cos(nx)$ pari, il prodotto $f(x)\cos(nx)$ è dispari, quindi $a_n=\dfrac1\pi\int_{-\pi}^\pi f(x)\cos(nx)\,dx=0$ per ogni $n\ge0$ (integrale di una funzione dispari su un intervallo simmetrico).

**Passo 2 — calcoliamo $b_n$.** $b_n=\dfrac1\pi\int_{-\pi}^\pi x\sin(nx)\,dx=\dfrac2\pi\int_0^\pi x\sin(nx)\,dx$ (essendo $x\sin(nx)$ pari, raddoppiamo l'integrale su metà dominio). Per parti (Modulo 5): $\int_0^\pi x\sin(nx)\,dx=\Big[-\dfrac{x\cos(nx)}n\Big]_0^\pi+\dfrac1n\int_0^\pi\cos(nx)\,dx=-\dfrac{\pi\cos(n\pi)}n+0=-\dfrac{\pi(-1)^n}n$ (essendo $\cos(n\pi)=(-1)^n$, e l'ultimo integrale nullo su un periodo intero se $n\ge1$). Dunque $b_n=\dfrac2\pi\cdot\Big(-\dfrac{\pi(-1)^n}n\Big)=\dfrac{-2(-1)^n}n=\dfrac{2(-1)^{n+1}}n$.

**Passo 3 — scriviamo la serie.** $f(x)\sim\sum_{n=1}^\infty\dfrac{2(-1)^{n+1}}n\sin(nx)=2\Big(\sin x-\dfrac{\sin2x}2+\dfrac{\sin3x}3-\cdots\Big)$.

**Passo 4 — applichiamo Parseval.** $\|f\|_2^2=\int_{-\pi}^\pi x^2\,dx=\Big[\dfrac{x^3}3\Big]_{-\pi}^\pi=\dfrac{2\pi^3}3$. La formula di Parseval (teoria, con $a_0=0$): $\|f\|_2^2=\pi\sum_{n=1}^\infty b_n^2=\pi\sum_{n=1}^\infty\dfrac4{n^2}$. Uguagliando: $\dfrac{2\pi^3}3=4\pi\sum_{n=1}^\infty\dfrac1{n^2}$, quindi $\sum_{n=1}^\infty\dfrac1{n^2}=\dfrac{2\pi^3}{3\cdot4\pi}=\dfrac{\pi^2}6$.

**Conclusione:** $\sum_{n=1}^\infty1/n^2=\pi^2/6$ — il **problema di Basilea**, già usato senza dimostrazione nel notebook 14.8, ora dimostrato rigorosamente come conseguenza dell'identità di Parseval applicata alla funzione $f(x)=x$. **Punto metodologico:** questo è un esempio paradigmatico di come le serie di Fourier, tramite Parseval, trasformino il calcolo di somme di serie numeriche (spesso difficili con tecniche dirette) in un semplice calcolo di un integrale.

```python
import numpy as np

# Verifica numerica: somma parziale di sum(1/n^2) confrontata con pi^2/6
N = 100000
somma = sum(1/n**2 for n in range(1, N+1))
print(f"Somma parziale (N={N}): {somma:.8f}")
print(f"pi^2/6 = {np.pi**2/6:.8f}")

# Verifica dei coefficienti b_n per f(x)=x
def b_n(n):
    return 2 * (-1)**(n+1) / n

for n in [1, 2, 3, 4, 5]:
    print(f"b_{n} = {b_n(n):.4f}")
```

**1.** $f(x)=x^2$ è pari, quindi $b_n=0$ per ogni $n$ (esercizio 3). $a_0=\dfrac1\pi\int_{-\pi}^\pi x^2\,dx=\dfrac1\pi\cdot\dfrac{2\pi^3}3=\dfrac{2\pi^2}3$. Per $n\ge1$: $a_n=\dfrac2\pi\int_0^\pi x^2\cos(nx)\,dx$; integrando per parti due volte (Modulo 5): $\int_0^\pi x^2\cos(nx)\,dx=\Big[\dfrac{x^2\sin(nx)}n\Big]_0^\pi-\dfrac2n\int_0^\pi x\sin(nx)\,dx=0-\dfrac2n\cdot\Big(-\dfrac{\pi(-1)^n}n\Big)=\dfrac{2\pi(-1)^n}{n^2}$ (riusando il calcolo dell'esercizio svolto per $\int_0^\pi x\sin(nx)dx$). **2.** Parseval: $\|f\|_2^2=\int_{-\pi}^\pi x^4\,dx=\Big[\dfrac{x^5}5\Big]_{-\pi}^\pi=\dfrac{2\pi^5}5$. Dall'altro lato: $\|f\|_2^2=\dfrac{a_0^2}2\pi+\pi\sum_n a_n^2=\dfrac{(2\pi^2/3)^2}2\pi+\pi\sum_n\dfrac{16}{n^4}=\dfrac{2\pi^5}9+16\pi\sum_n\dfrac1{n^4}$. Uguagliando: $\dfrac{2\pi^5}5-\dfrac{2\pi^5}9=16\pi\sum_n1/n^4$, cioè $2\pi^5\Big(\dfrac15-\dfrac19\Big)=2\pi^5\cdot\dfrac4{45}=\dfrac{8\pi^5}{45}=16\pi\sum_n1/n^4$, quindi $\sum_{n=1}^\infty1/n^4=\dfrac{8\pi^5}{45\cdot16\pi}=\dfrac{\pi^4}{90}$.

**3.** Se $f$ è pari, $f(x)\sin(nx)$ è il prodotto di una funzione pari e una dispari, quindi **dispari**: $f(-x)\sin(-nx)=f(x)\cdot(-\sin(nx))=-f(x)\sin(nx)$. L'integrale di una funzione dispari su un intervallo simmetrico $(-\pi,\pi)$ è sempre $0$ (Modulo 5): $b_n=\dfrac1\pi\int_{-\pi}^\pi f(x)\sin(nx)\,dx=0$.

**4.** La disuguaglianza di Bessel (valida per **qualunque** sistema ortonormale, anche non completo) dice $\sum_{k=1}^N\langle f,e_k\rangle^2\le\|f\|_2^2$ per ogni $N$ finito (essendo la proiezione su un sottospazio di dimensione finita sempre di norma minore o uguale all'originale, per Pitagora — un fatto dimostrabile direttamente senza assumere la completezza della base). ## Riepilogo finale

- Il sistema trigonometrico $\{1,\cos(nx),\sin(nx)\}$, opportunamente normalizzato, è una base ortonormale **completa** di $L^2(\mathbb T)$: la serie di Fourier è la proiezione ortogonale su questa base, un caso speciale della teoria generale degli spazi di Hilbert (notebook 14.3).
- La convergenza in norma $L^2$ è **sempre** garantita; la convergenza puntuale richiede ipotesi di regolarità aggiuntive (teorema di Dirichlet).
- L'**identità di Parseval** collega la norma $L^2$ di $f$ alla somma dei quadrati dei coefficienti di Fourier: uno strumento potente per calcolare somme di serie numeriche altrimenti difficili (il problema di Basilea, $\sum1/n^2=\pi^2/6$).
- La disuguaglianza di **Bessel** vale per ogni sistema ortonormale (anche incompleto); diventa uguaglianza (Parseval) solo con un sistema completo.
