# 15.4 — Equazione del calore

**Prerequisiti:** 15.3 Spazi di Fourier, 15.2 Trasformata di Fourier (rigorosa), 8.1 EDO — introduzione.


## Perché ci interessa

Fino a questo punto del corso abbiamo studiato equazioni **differenziali ordinarie** (Modulo 8): equazioni che coinvolgono derivate rispetto a **una sola** variabile. L'**equazione del calore** $\partial_tu=\partial_x^2u$ è la prima **equazione alle derivate parziali** (PDE) del corso: descrive come una distribuzione di temperatura $u(x,t)$ evolve nel tempo, diffondendosi nello spazio. È l'esempio guida per l'intera famiglia delle equazioni **paraboliche** (notebook 15.9), e la trasformata di Fourier (notebook 15.2) — che scambia derivazione in $x$ con moltiplicazione — la trasforma in una semplice **famiglia di equazioni differenziali ordinarie**, una per ogni frequenza $\xi$, risolvibili con gli strumenti già noti dal Modulo 8.

## Teoria, passo dopo passo

### Il problema di Cauchy per l'equazione del calore

Cerchiamo $u(x,t)$, per $x\in\mathbb R$, $t>0$, soluzione di $\partial_tu(x,t)=\partial_x^2u(x,t)$, con dato iniziale $u(x,0)=u_0(x)$ (la temperatura al tempo $t=0$).

### Risoluzione tramite trasformata di Fourier

Applichiamo la trasformata di Fourier in $x$ (notebook 15.2) a entrambi i membri, trattando $t$ come parametro: $\partial_t\hat u(\xi,t)=\widehat{\partial_x^2u}(\xi,t)=-4\pi^2\xi^2\hat u(\xi,t)$ (usando la proprietà derivazione↔moltiplicazione due volte, notebook 15.2, e scambiando $\partial_t$ con la trasformata in $x$ — lecito sotto ipotesi di regolarità che diamo per scontate).

Per ogni $\xi$ fissato, questa è un'**equazione differenziale ordinaria lineare del primo ordine** in $t$ (notebook 8.2, con $\hat u(\xi,t)$ come funzione incognita di $t$): $\partial_t\hat u=-4\pi^2\xi^2\hat u$, con soluzione (notebook 8.2, equazione a variabili separabili, o direttamente notebook 8.4 per equazioni lineari a coefficienti costanti) $\hat u(\xi,t)=\hat u(\xi,0)e^{-4\pi^2\xi^2t}=\hat u_0(\xi)e^{-4\pi^2\xi^2t}$.

### Il nucleo del calore

Antitrasformando (notebook 15.2), $u(\cdot,t)$ è il prodotto (in frequenza) di $\hat u_0$ con $e^{-4\pi^2\xi^2t}$ — che, per il teorema di convoluzione (una proprietà della trasformata di Fourier che scambia prodotti puntuali e convoluzioni, enunciata qui senza dimostrazione ma naturale conseguenza di Fubini, notebook 13.9), corrisponde a una **convoluzione** nello spazio fisico: $u(x,t)=(u_0*K_t)(x)=\int_{\mathbb R}u_0(y)K_t(x-y)\,dy$, dove $K_t$ è l'antitrasformata di $e^{-4\pi^2\xi^2t}$.

**Calcolo esplicito di $K_t$.** Riconosciamo $e^{-4\pi^2\xi^2t}$ come una gaussiana in $\xi$ (di "larghezza" che dipende da $t$): riscalando il risultato del notebook 15.2 (dove $\widehat{e^{-\pi x^2}}=e^{-\pi\xi^2}$) con un cambio di variabile, si ottiene $K_t(x)=\dfrac1{\sqrt{4\pi t}}e^{-x^2/(4t)}$ — il **nucleo del calore** (o nucleo di Gauss-Weierstrass).

### Proprietà del nucleo del calore

$K_t$ è essa stessa una gaussiana (di area totale $1$ per ogni $t$: $\int K_t\,dx=1$, verificabile con la sostituzione standard per l'integrale gaussiano), che si **appiattisce** al crescere di $t$ (la "larghezza" $\sqrt{4t}$ cresce) mantenendo l'area unitaria — esattamente il comportamento fisico atteso: il calore si diffonde, l'energia totale si conserva ma si spalma su una regione sempre più ampia.

**Regolarizzazione istantanea.** Per $t>0$, $K_t$ è di classe $C^\infty$ (essendo una gaussiana), quindi $u(\cdot,t)=u_0*K_t$ è **automaticamente liscia per ogni $t>0$**, **indipendentemente** dalla regolarità del dato iniziale $u_0$ (che potrebbe essere solo in $L^2$, o addirittura discontinuo): questo fenomeno, detto **effetto regolarizzante** dell'equazione del calore, è drasticamente diverso dal comportamento delle equazioni **iperboliche** (notebook 15.10, come l'equazione delle onde), dove le singolarità del dato iniziale **persistono** nel tempo.

### Buona posizione in $H^s$

**Teorema (enunciato).** Se $u_0\in H^s(\mathbb R)$ (notebook 15.3), esiste un'unica soluzione $u(\cdot,t)\in H^s(\mathbb R)$ per ogni $t\ge0$, con $\|u(\cdot,t)\|_{H^s}\le\|u_0\|_{H^s}$ per ogni $t\ge0$ (la norma $H^s$ non cresce nel tempo).

**Dimostrazione (schema).** $\|u(\cdot,t)\|_{H^s}^2=\int(1+\xi^2)^s|\hat u(\xi,t)|^2\,d\xi=\int(1+\xi^2)^s|\hat u_0(\xi)|^2e^{-8\pi^2\xi^2t}\,d\xi\le\int(1+\xi^2)^s|\hat u_0(\xi)|^2\,d\xi=\|u_0\|_{H^s}^2$ (essendo $e^{-8\pi^2\xi^2t}\le1$ per $t\ge0$, un fattore che *smorza* invece di amplificare). $\blacksquare$ Questo risultato di **stabilità** — piccole perturbazioni del dato iniziale producono piccole perturbazioni della soluzione, uniformemente nel tempo — è il prototipo della nozione di **buona posizione** che affronteremo sistematicamente nei notebook 15.8-15.9 sulle PDE paraboliche in generale.

```python
import numpy as np
import matplotlib.pyplot as plt

def K_t(x, t):
    return 1/np.sqrt(4*np.pi*t) * np.exp(-x**2/(4*t))

x = np.linspace(-5, 5, 500)
fig, ax = plt.subplots(figsize=(8,5))
for t in [0.05, 0.2, 0.5, 1.5]:
    ax.plot(x, K_t(x, t), label=f"K_t, t={t}")
ax.legend()
ax.set_title("Nucleo del calore: si appiattisce nel tempo, area sempre = 1")
plt.tight_layout()
plt.show()

for t in [0.05, 0.2, 0.5, 1.5]:
    area = np.trapz(K_t(x, t), x)
    print(f"t={t}: area sotto K_t = {area:.5f} (atteso: 1)")

# Verifica della soluzione: dato iniziale un'onda quadra approssimata, evoluta nel tempo (convoluzione numerica)
def u0(x):
    return np.where(np.abs(x) < 1, 1.0, 0.0)  # blocco rettangolare, discontinuo

x_fine = np.linspace(-10, 10, 2000)
dx = x_fine[1] - x_fine[0]

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x_fine, u0(x_fine), 'k--', label='u0 (dato iniziale, discontinuo)')
for t in [0.01, 0.1, 0.5]:
    kernel = K_t(x_fine, t)
    u_t = np.convolve(u0(x_fine), kernel, mode='same') * dx
    ax.plot(x_fine, u_t, label=f"u(.,t={t}) (liscia!)")
ax.legend()
ax.set_title("Effetto regolarizzante: dato discontinuo -> soluzione liscia per t>0")
plt.tight_layout()
plt.show()
```

## Esercizio svolto e commentato

**Problema:** verifica direttamente (senza passare dalla trasformata di Fourier) che $u(x,t)=K_t(x)=\dfrac1{\sqrt{4\pi t}}e^{-x^2/(4t)}$ soddisfa l'equazione del calore $\partial_tu=\partial_x^2u$ per $t>0$.

**Ragionamento:**

**Passo 1 — calcoliamo $\partial_xu$.** $u(x,t)=(4\pi t)^{-1/2}e^{-x^2/(4t)}$. Per la regola della catena (Modulo 5): $\partial_xu=(4\pi t)^{-1/2}e^{-x^2/(4t)}\cdot\Big(-\dfrac{2x}{4t}\Big)=-\dfrac{x}{2t}u$.

**Passo 2 — calcoliamo $\partial_x^2u$.** Derivando ancora, con la regola del prodotto (Modulo 5): $\partial_x^2u=-\dfrac1{2t}u-\dfrac{x}{2t}\partial_xu=-\dfrac1{2t}u-\dfrac{x}{2t}\cdot\Big(-\dfrac x{2t}u\Big)=-\dfrac1{2t}u+\dfrac{x^2}{4t^2}u=u\Big(\dfrac{x^2}{4t^2}-\dfrac1{2t}\Big)$.

**Passo 3 — calcoliamo $\partial_tu$.** Scriviamo $u=(4\pi)^{-1/2}t^{-1/2}e^{-x^2/(4t)}$. Per la regola del prodotto e della catena: $\partial_tu=(4\pi)^{-1/2}\Big[-\dfrac12t^{-3/2}e^{-x^2/(4t)}+t^{-1/2}e^{-x^2/(4t)}\cdot\dfrac{x^2}{4t^2}\Big]=u\Big(-\dfrac1{2t}+\dfrac{x^2}{4t^2}\Big)$ (raccogliendo $u=(4\pi)^{-1/2}t^{-1/2}e^{-x^2/(4t)}$ da entrambi i termini).

**Passo 4 — confrontiamo.** $\partial_tu=u\Big(\dfrac{x^2}{4t^2}-\dfrac1{2t}\Big)=\partial_x^2u$ (identico al risultato del passo 2).

**Conclusione:** $K_t$ soddisfa esattamente l'equazione del calore per $t>0$, verificando direttamente per calcolo diretto ciò che la trasformata di Fourier aveva prodotto come antitrasformata. **Nota:** questa verifica diretta, pur essendo un utile controllo di consistenza, **non** dimostra da sola che $K_t$ sia la soluzione "giusta" con dato iniziale $\delta_0$ (la Delta di Dirac, un oggetto che esula dalla teoria delle funzioni classiche trattata qui) — è la costruzione tramite trasformata di Fourier, con il passaggio al limite $\hat u(\xi,0)=\hat u_0(\xi)$, a garantire rigorosamente che $u_0*K_t\to u_0$ quando $t\to0^+$ (in un senso $L^2$ o puntuale, a seconda della regolarità di $u_0$).

```python
import sympy as sp

x, t = sp.symbols('x t', positive=True, real=True)
u = 1/sp.sqrt(4*sp.pi*t) * sp.exp(-x**2/(4*t))

du_dt = sp.diff(u, t)
d2u_dx2 = sp.diff(u, x, 2)

differenza = sp.simplify(du_dt - d2u_dx2)
print("d(u)/dt - d^2(u)/dx^2 =", differenza)
print("L'equazione del calore e' soddisfatta esattamente" if differenza == 0 else "ERRORE")
```

## Esercizi da risolvere

1. Verifica che $\int_{\mathbb R}K_t(x)\,dx=1$ per ogni $t>0$ (suggerimento: usa la sostituzione $y=x/\sqrt{4t}$ per ricondurti all'integrale gaussiano standard $\int e^{-y^2}\,dy=\sqrt\pi$).
2. Dimostra che se $u_0\ge0$ (temperatura iniziale non negativa), allora $u(x,t)\ge0$ per ogni $x,t$ (suggerimento: usa la formula di convoluzione e il fatto che $K_t>0$ ovunque).
3. Dimostra il **principio del massimo** in forma debole: $\sup_xu(x,t)\le\sup_xu_0(x)$ per ogni $t>0$ (suggerimento: usa $u(x,t)=\int u_0(y)K_t(x-y)\,dy\le\sup_yu_0(y)\cdot\int K_t(x-y)\,dy$ e l'esercizio 1).
4. *(più impegnativo)* Usa la formula $\hat u(\xi,t)=\hat u_0(\xi)e^{-4\pi^2\xi^2t}$ per dimostrare che, per $t>0$, $u(\cdot,t)\in H^s(\mathbb R)$ per **ogni** $s\ge0$, anche se $u_0\in L^2(\mathbb R)$ soltanto (cioè $u_0\in H^0$) — una dimostrazione rigorosa dell'effetto regolarizzante enunciato nella teoria (suggerimento: stima $(1+\xi^2)^se^{-8\pi^2\xi^2t}$ con una costante che dipende da $s,t$ ma non da $\xi$, usando che l'esponenziale domina ogni potenza).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Con $y=x/\sqrt{4t}$ (quindi $dx=\sqrt{4t}\,dy$): $\int_{\mathbb R}K_t(x)\,dx=\int_{\mathbb R}\dfrac1{\sqrt{4\pi t}}e^{-x^2/(4t)}\,dx=\dfrac1{\sqrt{4\pi t}}\int_{\mathbb R}e^{-y^2}\sqrt{4t}\,dy=\dfrac{\sqrt{4t}}{\sqrt{4\pi t}}\int e^{-y^2}\,dy=\dfrac1{\sqrt\pi}\cdot\sqrt\pi=1$ (usando l'integrale gaussiano standard $\int_{\mathbb R}e^{-y^2}\,dy=\sqrt\pi$, dimostrato nel notebook 7.15 con il trucco delle coordinate polari).

**2.** $u(x,t)=\int_{\mathbb R}u_0(y)K_t(x-y)\,dy$: essendo $u_0(y)\ge0$ per ipotesi e $K_t(x-y)=\dfrac1{\sqrt{4\pi t}}e^{-(x-y)^2/(4t)}>0$ per ogni $x,y$ (un esponenziale è sempre positivo), l'integranda $u_0(y)K_t(x-y)\ge0$ per ogni $y$: l'integrale di una funzione non negativa è non negativo (notebook 13.4, o direttamente Modulo 5), quindi $u(x,t)\ge0$.

**3.** Sia $M=\sup_yu_0(y)$. $u(x,t)=\int u_0(y)K_t(x-y)\,dy\le\int M\cdot K_t(x-y)\,dy=M\int K_t(x-y)\,dy=M\cdot1=M$ (usando $u_0(y)\le M$ per ogni $y$, poi la monotonia dell'integrale, poi l'esercizio 1 con un cambio di variabile $z=x-y$ che non altera l'integrale su tutto $\mathbb R$). Dunque $u(x,t)\le M=\sup_yu_0(y)$ per ogni $x$, cioè $\sup_xu(x,t)\le\sup_xu_0(x)$.

**4.** $\|u(\cdot,t)\|_{H^s}^2=\int(1+\xi^2)^s|\hat u_0(\xi)|^2e^{-8\pi^2\xi^2t}\,d\xi$. Per $t>0$ fissato, la funzione $\xi\mapsto(1+\xi^2)^se^{-8\pi^2\xi^2t}$ è continua e tende a $0$ per $|\xi|\to\infty$ (l'esponenziale decrescente domina la crescita polinomiale $(1+\xi^2)^s$, per qualunque $s$ fissato, notebook 4.2/10.10 generalizzato): dunque è **limitata** su tutto $\mathbb R$, diciamo da una costante $C_{s,t}$ (il suo massimo, raggiunto essendo continua e tendente a $0$ agli estremi — un argomento simile al teorema di Weierstrass, notebook 10.12, esteso a $\mathbb R$ tramite il decadimento all'infinito). Dunque $\|u(\cdot,t)\|_{H^s}^2\le C_{s,t}\int|\hat u_0(\xi)|^2\,d\xi=C_{s,t}\|u_0\|_2^2<\infty$ (essendo $u_0\in L^2$ per ipotesi): $u(\cdot,t)\in H^s(\mathbb R)$ per ogni $s\ge0$, anche se $u_0$ sta solo in $H^0=L^2$.

## Riepilogo finale

- L'**equazione del calore** $\partial_tu=\partial_x^2u$, trasformata secondo Fourier in $x$, diventa una famiglia di equazioni differenziali ordinarie del primo ordine (una per frequenza), con soluzione esplicita $\hat u(\xi,t)=\hat u_0(\xi)e^{-4\pi^2\xi^2t}$.
- La soluzione si scrive come convoluzione col **nucleo del calore** $K_t(x)=(4\pi t)^{-1/2}e^{-x^2/(4t)}$, una gaussiana che si appiattisce nel tempo mantenendo area unitaria.
- **Effetto regolarizzante**: per $t>0$, la soluzione è sempre liscia (in ogni $H^s$), indipendentemente dalla regolarità del dato iniziale — dimostrato rigorosamente stimando il decadimento esponenziale in frequenza.
- La norma $H^s$ della soluzione non cresce nel tempo: un primo esempio di **buona posizione** (stabilità rispetto al dato iniziale), il concetto che guiderà lo studio sistematico delle PDE nei prossimi notebook.
