# 8.4 — Equazioni a coefficienti costanti

**Prerequisiti:** 8.3 Equazioni lineari, 1.6 Equazioni di secondo grado, 5.9 Teoremi fondamentali della derivazione.


## Perché ci interessa

L'equazione $y''+\omega^2y=0$ descrive l'**oscillatore armonico**: una molla, un pendolo per piccole oscillazioni, un circuito LC — uno dei modelli più importanti della fisica. È un'equazione del **secondo ordine**, lineare, a **coefficienti costanti**. In questo notebook sviluppiamo un metodo sorprendentemente semplice per risolvere l'intera classe $ay''+by'+cy=0$: si tenta una soluzione esponenziale $y=e^{\lambda t}$, e il problema si riduce a un'**equazione algebrica di secondo grado** — esattamente quella del notebook 1.6.

## Teoria, passo dopo passo

### L'equazione caratteristica

Consideriamo l'equazione omogenea a coefficienti costanti $ay''+by'+cy=0$ (con $a\ne0$). Proviamo una soluzione della forma $y(t)=e^{\lambda t}$: allora $y'=\lambda e^{\lambda t}$, $y''=\lambda^2e^{\lambda t}$ (derivando l'esponenziale, notebook 5.9). Sostituendo:

$$a\lambda^2e^{\lambda t}+b\lambda e^{\lambda t}+ce^{\lambda t}=0 \quad\Longrightarrow\quad e^{\lambda t}(a\lambda^2+b\lambda+c)=0$$

Poiché $e^{\lambda t}\ne0$ sempre, deve essere $a\lambda^2+b\lambda+c=0$: l'**equazione caratteristica**, un'equazione di secondo grado (notebook 1.6) nell'incognita $\lambda$. Ogni radice $\lambda$ dà una soluzione $e^{\lambda t}$.

### Caso 1: due radici reali distinte

Se il discriminante $\Delta=b^2-4ac>0$, ci sono due radici reali distinte $\lambda_1\ne\lambda_2$ (notebook 1.6). La soluzione generale è:

$$y(t) = C_1e^{\lambda_1t}+C_2e^{\lambda_2t}$$

(si può verificare che ogni combinazione lineare di soluzioni di un'equazione lineare omogenea è ancora soluzione — una conseguenza della linearità della derivata, notebook 5.9 — e che, con due radici distinte, questa famiglia a due parametri esaurisce tutte le soluzioni).

### Caso 2: radice doppia

Se $\Delta=0$, c'è un'unica radice $\lambda=-b/(2a)$ (notebook 1.6), con molteplicità 2. Una sola soluzione esponenziale $e^{\lambda t}$ non basta (servono due soluzioni indipendenti per una soluzione generale a due parametri, come nel Caso 1). Si dimostra (verificabile per sostituzione diretta) che anche $t\,e^{\lambda t}$ risolve l'equazione in questo caso:

$$y(t) = (C_1+C_2t)e^{\lambda t}$$

### Caso 3: radici complesse coniugate

Se $\Delta<0$, le radici sono complesse coniugate $\lambda=\alpha\pm i\beta$ (con $i^2=-1$, i numeri complessi saranno formalizzati nel Modulo 9; qui ne usiamo solo la formula risolutiva estesa). Usando la **formula di Eulero** $e^{i\beta t}=\cos(\beta t)+i\sin(\beta t)$ (anticipata qui, dimostrata nel Modulo 9), si combinano le due soluzioni complesse in due soluzioni **reali**:

$$y(t) = e^{\alpha t}\left(C_1\cos(\beta t)+C_2\sin(\beta t)\right)$$

Per l'oscillatore armonico $y''+\omega^2y=0$: $a=1,b=0,c=\omega^2$, discriminante $-4\omega^2<0$, radici $\lambda=\pm i\omega$ (cioè $\alpha=0,\beta=\omega$): $y(t)=C_1\cos(\omega t)+C_2\sin(\omega t)$ — esattamente la soluzione verificata nell'esercizio 1 del notebook 8.1!

### Equazioni non omogenee: soluzione particolare + omogenea

Per $ay''+by'+cy=f(t)$ (non omogenea), la soluzione generale è $y=y_p+y_h$ (esattamente il principio del notebook 8.3), dove $y_h$ è la soluzione generale dell'omogenea associata (i tre casi sopra) e $y_p$ è **una qualsiasi** soluzione particolare, spesso trovata per **tentativi** (metodo dei coefficienti indeterminati): se $f(t)$ è un polinomio, si prova $y_p$ polinomiale; se $f(t)=e^{kt}$, si prova $y_p=Ae^{kt}$ (a meno che $k$ non sia già radice caratteristica, un caso di **risonanza** che richiede un fattore $t$ in più, analogo al Caso 2).

````python
import sympy as sp

t = sp.symbols('t', real=True)
lam = sp.symbols('lambda')
y_func = sp.Function('y')

# Caso 1: radici reali distinte, y'' - 3y' + 2y = 0
a, b, c = 1, -3, 2
eq_caratteristica = a*lam**2 + b*lam + c
radici = sp.solve(eq_caratteristica, lam)
print("Radici caratteristiche (caso 1):", radici)

eq1 = sp.Eq(y_func(t).diff(t, 2) - 3*y_func(t).diff(t) + 2*y_func(t), 0)
print("Soluzione generale (caso 1):", sp.dsolve(eq1, y_func(t)))

# Caso 3: oscillatore armonico y'' + omega^2 y = 0
omega = sp.symbols('omega', positive=True)
eq3 = sp.Eq(y_func(t).diff(t, 2) + omega**2 * y_func(t), 0)
print("\nSoluzione generale (oscillatore armonico):", sp.dsolve(eq3, y_func(t)))

# Non omogenea: y'' + y = t (usando il metodo dei coefficienti indeterminati automatico di SymPy)
eq_non_omog = sp.Eq(y_func(t).diff(t, 2) + y_func(t), t)
print("\nSoluzione (non omogenea, y''+y=t):", sp.dsolve(eq_non_omog, y_func(t)))
````

## Esercizio svolto e commentato

**Problema:** risolvi il problema di Cauchy $\begin{cases}y''-4y'+4y=0\\y(0)=1,\ y'(0)=0\end{cases}$.

**Ragionamento:**

**Passo 1 — scriviamo l'equazione caratteristica**: $\lambda^2-4\lambda+4=0$.

**Passo 2 — calcoliamo il discriminante** (notebook 1.6): $\Delta=16-16=0$. **Radice doppia**: $\lambda=\dfrac{4}{2}=2$.

**Passo 3 — applichiamo la formula del Caso 2**: $y(t)=(C_1+C_2t)e^{2t}$.

**Passo 4 — calcoliamo $y'(t)$**, necessaria per la seconda condizione iniziale (regola del prodotto, notebook 5.9): $y'(t)=C_2e^{2t}+(C_1+C_2t)\times2e^{2t}=(2C_1+C_2+2C_2t)e^{2t}$.

**Passo 5 — imponiamo $y(0)=1$**: $y(0)=(C_1+0)\times1=C_1=1$.

**Passo 6 — imponiamo $y'(0)=0$**: $y'(0)=(2C_1+C_2+0)\times1=2\times1+C_2=0$, quindi $C_2=-2$.

**Conclusione:** $y(t)=(1-2t)e^{2t}$.

**Perché un'equazione del secondo ordine richiede **due** condizioni iniziali (qui $y(0)$ e $y'(0)$), non una sola come per il primo ordine:** la soluzione generale di un'equazione del secondo ordine ha **due** costanti arbitrarie ($C_1,C_2$), riflettendo il fatto che due integrazioni "consecutive" sono coinvolte nel processo risolutivo (concettualmente, come integrare due volte per passare da $y''$ a $y$); servono quindi due equazioni indipendenti — tipicamente valore e derivata nello stesso istante — per determinare univocamente entrambe le costanti, esattamente come, in meccanica, conoscere posizione **e** velocità iniziali di un oggetto determina completamente il suo moto futuro (l'equazione del moto essendo tipicamente del secondo ordine, legata all'accelerazione).

````python
import sympy as sp

t = sp.symbols('t', real=True)
y_func = sp.Function('y')

eq = sp.Eq(y_func(t).diff(t, 2) - 4*y_func(t).diff(t) + 4*y_func(t), 0)
soluzione = sp.dsolve(eq, y_func(t), ics={y_func(0): 1, y_func(t).diff(t).subs(t, 0): 0})
print("Soluzione:", soluzione)
````

## Riepilogo finale

- Per $ay''+by'+cy=0$, si tenta $y=e^{\lambda t}$, ottenendo l'**equazione caratteristica** $a\lambda^2+b\lambda+c=0$ (notebook 1.6).
- Tre casi, a seconda del discriminante: radici reali distinte ($C_1e^{\lambda_1t}+C_2e^{\lambda_2t}$), radice doppia ($(C_1+C_2t)e^{\lambda t}$), radici complesse coniugate ($e^{\alpha t}(C_1\cos\beta t+C_2\sin\beta t)$).
- Servono **due** condizioni iniziali ($y(0)$ e $y'(0)$) per determinare univocamente $C_1,C_2$.
- Per equazioni non omogenee, si cerca una soluzione particolare per tentativi, con attenzione al fenomeno della **risonanza** quando il termine noto "assomiglia" a una soluzione dell'omogenea.
