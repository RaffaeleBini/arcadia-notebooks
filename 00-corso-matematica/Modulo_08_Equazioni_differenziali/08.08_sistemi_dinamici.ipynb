## Perché ci interessa

Il mondo reale raramente è **lineare**: la crescita di una popolazione preda-predatore, la dinamica di un pendolo non limitato a piccole oscillazioni, le reazioni chimiche autocatalitiche seguono equazioni $\mathbf x'=\mathbf F(\mathbf x)$ con $\mathbf F$ **non lineare**. Risolvere esplicitamente questi sistemi è spesso impossibile. Ma vicino a un **equilibrio**, possiamo approssimare $\mathbf F$ con la sua parte lineare (la formula di Taylor del notebook 5.14, applicata qui a funzioni vettoriali) e applicare tutta la teoria di stabilità del notebook 8.7 al sistema linearizzato — un'idea potentissima, che estende l'analisi lineare a un'enorme classe di problemi realistici.

## Teoria, passo dopo passo

### Sistemi non lineari e punti di equilibrio

Un **sistema dinamico** (autonomo, cioè senza dipendenza esplicita da $t$) è $\mathbf x'=\mathbf F(\mathbf x)$, con $\mathbf F:\mathbb R^n\to\mathbb R^n$ un campo vettoriale (notebook 7.20), non necessariamente lineare. Un **equilibrio** è un punto $\mathbf x^*$ dove $\mathbf F(\mathbf x^*)=\mathbf 0$: se il sistema parte esattamente lì, resta lì per sempre (la velocità è nulla).

### Linearizzazione: la matrice Jacobiana

Vicino a un equilibrio $\mathbf x^*$, scriviamo $\mathbf x=\mathbf x^*+\mathbf h$ (una piccola perturbazione). Per la formula di Taylor al primo ordine, applicata a ciascuna componente di $\mathbf F$ (notebook 7.8, generalizzata componente per componente):

$$\mathbf F(\mathbf x^*+\mathbf h) \approx \mathbf F(\mathbf x^*) + J_{\mathbf F}(\mathbf x^*)\,\mathbf h = J_{\mathbf F}(\mathbf x^*)\,\mathbf h$$

(usando $\mathbf F(\mathbf x^*)=\mathbf 0$), dove $J_{\mathbf F}$ è la **matrice Jacobiana** di $\mathbf F$ (notebook 7.15: la matrice delle derivate parziali di ciascuna componente rispetto a ciascuna variabile). Poiché $\mathbf x'=\mathbf h'$ (essendo $\mathbf x^*$ costante), il sistema vicino all'equilibrio si approssima con:

$$\mathbf h' \approx J_{\mathbf F}(\mathbf x^*)\,\mathbf h$$

un sistema **lineare** nella perturbazione $\mathbf h$, con matrice dei coefficienti $A=J_{\mathbf F}(\mathbf x^*)$! Possiamo quindi applicare **direttamente** la teoria di stabilità del notebook 8.7 a questo sistema linearizzato.

### Il teorema di Hartman-Grobman (enunciato informale)

> Se gli autovalori di $J_{\mathbf F}(\mathbf x^*)$ hanno **tutti** parte reale non nulla (un equilibrio **iperbolico**), allora il comportamento qualitativo del sistema non lineare vicino a $\mathbf x^*$ è "lo stesso" di quello del sistema linearizzato: se la linearizzazione predice un nodo stabile, una sella, o una spirale, il sistema originale si comporta effettivamente così vicino all'equilibrio.

Questo giustifica rigorosamente l'uso della linearizzazione: **non** è solo un'approssimazione grossolana, ma cattura fedelmente il comportamento locale (tranne nel caso delicato di autovalori con parte reale esattamente nulla — i "centri" — dove i termini non lineari, trascurati dalla linearizzazione, possono cambiare qualitativamente il comportamento).

### Esempio: il pendolo

L'equazione del pendolo (non linearizzata) è $\theta''=-\dfrac g\ell\sin\theta$. In forma di sistema, con $x_1=\theta,x_2=\theta'$: $\mathbf F(x_1,x_2)=\left(x_2,\,-\dfrac g\ell\sin x_1\right)$. Gli equilibri: $x_2=0$ e $\sin x_1=0$, cioè $x_1=0,\pi,2\pi,\dots$ — il pendolo verso il basso ($x_1=0$) e verso l'alto ($x_1=\pi$). La Jacobiana è $J_{\mathbf F}=\begin{pmatrix}0&1\\-\frac g\ell\cos x_1&0\end{pmatrix}$. In $x_1=0$: $J=\begin{pmatrix}0&1\\-g/\ell&0\end{pmatrix}$, con $\det=g/\ell>0$, $\operatorname{tr}=0$: un **centro** (oscillazioni, coerente con l'intuizione fisica — il pendolo verso il basso oscilla). In $x_1=\pi$: $\cos\pi=-1$, quindi $J=\begin{pmatrix}0&1\\g/\ell&0\end{pmatrix}$, con $\det=-g/\ell<0$: una **sella**, instabile — coerente con l'intuizione che il pendolo in equilibrio verso l'alto è instabile, basta una minima perturbazione per farlo cadere.

````python
import sympy as sp

x1, x2, g, l = sp.symbols('x1 x2 g ell', positive=True)

F = sp.Matrix([x2, -(g/l) * sp.sin(x1)])
J = F.jacobian([x1, x2])
print("Jacobiana generale:\n", J)

# Equilibrio 1: pendolo verso il basso (x1=0)
J_basso = J.subs({x1: 0, x2: 0})
print("\nJacobiana in (0,0):\n", J_basso)
print("tr =", J_basso.trace(), " det =", J_basso.det(), "-> centro (oscillazioni)")

# Equilibrio 2: pendolo verso l'alto (x1=pi)
J_alto = J.subs({x1: sp.pi, x2: 0})
print("\nJacobiana in (pi,0):\n", J_alto)
print("tr =", J_alto.trace(), " det =", J_alto.det(), "-> sella (instabile)")
````

## Esercizio svolto e commentato

**Problema:** trova gli equilibri del sistema preda-predatore (modello di Lotka-Volterra semplificato) $\begin{cases}x'=x(2-y)\\y'=y(x-1)\end{cases}$ e classifica la loro stabilità.

**Ragionamento:**

**Passo 1 — troviamo gli equilibri**, risolvendo $\mathbf F(x,y)=\mathbf 0$: $x(2-y)=0$ dà $x=0$ oppure $y=2$; $y(x-1)=0$ dà $y=0$ oppure $x=1$. Combinando: $(x,y)=(0,0)$ (entrambe le prime condizioni con $x=0,y=0$) oppure $(x,y)=(1,2)$ ($y=2$ e $x=1$).

**Passo 2 — calcoliamo la Jacobiana generale**: $J=\begin{pmatrix}\partial(x(2-y))/\partial x&\partial(x(2-y))/\partial y\\\partial(y(x-1))/\partial x&\partial(y(x-1))/\partial y\end{pmatrix}=\begin{pmatrix}2-y&-x\\y&x-1\end{pmatrix}$.

**Passo 3 — valutiamo in $(0,0)$**: $J(0,0)=\begin{pmatrix}2&0\\0&-1\end{pmatrix}$. Autovalori immediati (diagonale): $2,-1$, segni opposti: **sella**, instabile.

**Passo 4 — valutiamo in $(1,2)$**: $J(1,2)=\begin{pmatrix}0&-1\\2&0\end{pmatrix}$. $\operatorname{tr}=0$, $\det=0-(-2)=2>0$: **centro** (oscillazioni, il caso delicato menzionato nella teoria).

**Conclusione:** l'equilibrio "banale" $(0,0)$ (estinzione di entrambe le specie) è instabile (una sella): perturbazioni tipiche allontanano il sistema da questo stato. L'equilibrio $(1,2)$ (coesistenza) è un centro: la linearizzazione predice oscillazioni, coerente con il celebre comportamento ciclico delle popolazioni preda-predatore osservato in natura (anche se, essendo un centro, la teoria di Hartman-Grobman non garantisce che il sistema **non lineare** si comporti esattamente come oscillazioni pure — servirebbe un'analisi più fine, oltre lo scopo di questo notebook, per confermarlo rigorosamente in questo caso specifico).

**Perché il caso $(1,2)$ è un esempio calzante del limite della linearizzazione menzionato nella teoria:** essendo un centro (autovalori con parte reale esattamente nulla), il teorema di Hartman-Grobman **non si applica**: la linearizzazione suggerisce oscillazioni, ma il comportamento del sistema non lineare originale potrebbe, in linea di principio, essere leggermente diverso (una spirale molto lenta, per esempio) — un promemoria che la linearizzazione è uno strumento potente ma con un'eccezione importante da tenere sempre a mente.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
F = sp.Matrix([x * (2 - y), y * (x - 1)])

equilibri = sp.solve([F[0], F[1]], [x, y])
print("Equilibri:", equilibri)

J = F.jacobian([x, y])
print("\nJacobiana generale:\n", J)

for eq in equilibri:
    J_eq = J.subs({x: eq[0], y: eq[1]})
    print(f"\nJacobiana in {eq}:\n", J_eq, " tr=", J_eq.trace(), " det=", J_eq.det())
````

## Riepilogo finale

- Per sistemi non lineari $\mathbf x'=\mathbf F(\mathbf x)$, un **equilibrio** $\mathbf x^*$ soddisfa $\mathbf F(\mathbf x^*)=\mathbf 0$.
- La **matrice Jacobiana** $J_{\mathbf F}(\mathbf x^*)$ (notebook 7.15) fornisce la linearizzazione del sistema vicino all'equilibrio.
- Il **teorema di Hartman-Grobman** garantisce che, per equilibri iperbolici (autovalori con parte reale non nulla), il comportamento qualitativo locale coincide con quello del sistema linearizzato.
- Il caso degli autovalori con parte reale nulla (centri) è delicato: la linearizzazione da sola non basta a determinare il comportamento del sistema non lineare originale.
