# 8.5 — Sistemi di equazioni differenziali

**Prerequisiti:** 8.4 Equazioni a coefficienti costanti, 6.1 Vettori, 6.10 Matrici.


## Perché ci interessa

Molti fenomeni coinvolgono **più** grandezze che cambiano nel tempo, influenzandosi a vicenda: le popolazioni di una preda e di un predatore, le concentrazioni di due sostanze chimiche che reagiscono, le correnti in un circuito con più maglie. Descrivere questi fenomeni richiede **più** equazioni differenziali, accoppiate tra loro. In questo notebook impariamo a scrivere questi sistemi in forma **vettoriale**, un passo che permette di riusare direttamente gli strumenti dell'algebra lineare (Modulo 6) per risolverli — l'argomento del prossimo notebook.

## Teoria, passo dopo passo

### Un sistema di due equazioni

Consideriamo due funzioni incognite $x(t),y(t)$ che soddisfano:

$$\begin{cases}x'(t) = ax(t)+by(t)\\ y'(t) = cx(t)+dy(t)\end{cases}$$

(un sistema **lineare** a coefficienti costanti, il caso più semplice ma già ricchissimo). Ogni equazione coinvolge **entrambe** le incognite: non si può risolvere una senza l'altra.

### Forma vettoriale

Introducendo il vettore $\mathbf x(t)=(x(t),y(t))\in\mathbb R^2$ (notebook 6.1) e la matrice dei coefficienti $A=\begin{pmatrix}a&b\\c&d\end{pmatrix}$ (notebook 6.10), il sistema si riscrive compattamente:

$$\mathbf x'(t) = A\mathbf x(t)$$

dove $\mathbf x'(t)=(x'(t),y'(t))$ (derivata componente per componente, come già visto per le curve parametriche, notebook 7.18). Questa è la stessa forma dell'equazione scalare $y'=ay$ (notebook 8.2), con lo scalare $a$ sostituito dalla matrice $A$ e la funzione $y$ sostituita dal vettore $\mathbf x$ — un'analogia che guiderà tutta la teoria di questo e del prossimo notebook.

### Riduzione di un'equazione di ordine superiore a un sistema del primo ordine

Un fatto notevole: **ogni** equazione lineare di ordine $n$ si può riscrivere come un sistema di $n$ equazioni del **primo** ordine. Per esempio, l'equazione del secondo ordine $y''+by'+cy=0$ (notebook 8.4) si riduce introducendo la variabile ausiliaria $v=y'$ (la "velocità"):

$$\begin{cases}y' = v\\ v' = -cy-bv\end{cases}$$

(la seconda equazione è semplicemente $y''=-cy-by'$ riscritta con $v$ al posto di $y'$). In forma vettoriale, con $\mathbf x=(y,v)$:

$$\mathbf x' = \begin{pmatrix}0&1\\-c&-b\end{pmatrix}\mathbf x$$

Questa riduzione — sempre possibile — significa che lo studio dei sistemi del primo ordine (che affronteremo nei prossimi notebook) **include** come caso particolare lo studio di equazioni di ordine qualsiasi: non perdiamo generalità concentrandoci sui sistemi $\mathbf x'=A\mathbf x$.

### Interpretazione geometrica: il piano delle fasi

Per un sistema $2\times2$, possiamo visualizzare le soluzioni $\mathbf x(t)=(x(t),y(t))$ come **traiettorie** nel piano $(x,y)$ (detto **piano delle fasi**), parametrizzate dal tempo $t$ — esattamente come le curve parametriche del notebook 7.18. In ogni punto del piano, il vettore $A\mathbf x$ indica la direzione e la velocità istantanea del moto: lo stesso concetto di campo di direzioni del notebook 8.1, ora in due dimensioni, dove il "campo vettoriale" (notebook 7.20) è dato semplicemente da $\mathbf x\mapsto A\mathbf x$, un'applicazione **lineare** (notebook 6.13).

````python
import numpy as np
import matplotlib.pyplot as plt

# Sistema x' = Ax, con A = [[0,1],[-4,0]] (equivalente a y''+4y=0, oscillatore armonico)
A = np.array([[0, 1], [-4, 0]])

def campo(x, y):
    v = A @ np.array([x, y])
    return v[0], v[1]

X = np.linspace(-2, 2, 20)
Y = np.linspace(-2, 2, 20)
Xg, Yg = np.meshgrid(X, Y)
U, V = campo(Xg, Yg)

fig, ax = plt.subplots(figsize=(7, 7))
ax.streamplot(Xg, Yg, U, V, color='gray', density=1.2)

# Sovrapponiamo una traiettoria: soluzione con x(0)=1, y(0)=0 (cioè y(0)=1, y'(0)=0 nell'equazione originale)
t = np.linspace(0, 10, 300)
omega = 2  # perché c=4=omega^2
x_t = np.cos(omega * t)
y_t = -omega * np.sin(omega * t)
ax.plot(x_t, y_t, color='red', linewidth=2)

ax.set_xlabel('x (posizione)')
ax.set_ylabel('y (velocità)')
ax.set_title("Piano delle fasi per x'=Ax (oscillatore armonico riscritto come sistema)")
ax.set_aspect('equal')
plt.show()

print("Matrice A:\n", A)
````

## Esercizio svolto e commentato

**Problema:** riscrivi l'equazione del terzo ordine $y'''-2y''+y'-2y=0$ come sistema del primo ordine.

**Ragionamento:**

**Passo 1 — introduciamo variabili ausiliarie** per ridurre l'ordine: poniamo $x_1=y$, $x_2=y'$, $x_3=y''$ (la funzione stessa, la sua derivata, e la sua derivata seconda).

**Passo 2 — scriviamo le equazioni "banali"** che collegano queste variabili tra loro: $x_1'=y'=x_2$ (per definizione di $x_2$). $x_2'=y''=x_3$ (per definizione di $x_3$).

**Passo 3 — riscriviamo l'equazione originale** in termini delle nuove variabili: da $y'''-2y''+y'-2y=0$, isoliamo $y'''=2y''-y'+2y$, cioè $x_3'=2x_3-x_2+2x_1$.

**Passo 4 — componiamo il sistema completo**:

$$\begin{cases}x_1'=x_2\\x_2'=x_3\\x_3'=2x_1-x_2+2x_3\end{cases}$$

**Passo 5 — scriviamo la forma matriciale**, con $\mathbf x=(x_1,x_2,x_3)$:

$$\mathbf x' = \begin{pmatrix}0&1&0\\0&0&1\\2&-1&2\end{pmatrix}\mathbf x$$

**Conclusione:** l'equazione del terzo ordine è equivalente al sistema $3\times3$ mostrato.

**Perché la struttura della matrice ottenuta (con una sottodiagonale di $1$ e l'ultima riga contenente i coefficienti dell'equazione originale, in ordine e con segni opportuni) è sempre la stessa per questo tipo di riduzione, indipendentemente dall'equazione di partenza:** questa matrice — chiamata **matrice compagna** dell'equazione — codifica direttamente i coefficienti dell'equazione originale nell'ultima riga (qui, $2,-1,2$, corrispondenti a come $y'''$ si esprime in funzione di $y,y',y''$), mentre le righe superiori si limitano a "trasportare" le definizioni ausiliarie ($x_1'=x_2$, $x_2'=x_3$); riconoscere questo schema generale permette di scrivere direttamente la forma matriciale per **qualsiasi** equazione lineare di ordine $n$, senza dover rifare il ragionamento da capo ogni volta.

````python
import sympy as sp

x1, x2, x3 = sp.symbols('x1 x2 x3')
A = sp.Matrix([[0, 1, 0], [0, 0, 1], [2, -1, 2]])
x = sp.Matrix([x1, x2, x3])

print("A*x =", (A * x).T)
print("\nQuesto corrisponde a: x1'=x2, x2'=x3, x3'=2x1-x2+2x3")

# Verifica: se y risolve y'''-2y''+y'-2y=0, allora x=(y,y',y'') risolve x'=Ax
t = sp.symbols('t')
y_func = sp.Function('y')
eq_originale = sp.Eq(y_func(t).diff(t, 3) - 2*y_func(t).diff(t, 2) + y_func(t).diff(t) - 2*y_func(t), 0)
print("\nEquazione originale:", eq_originale)
````

## Riepilogo finale

- Un **sistema di equazioni differenziali lineari** si scrive compattamente $\mathbf x'=A\mathbf x$, con $\mathbf x$ vettore delle incognite e $A$ matrice dei coefficienti (notebook 6.1, 6.10).
- **Ogni** equazione lineare di ordine $n$ si riduce a un sistema del primo ordine di $n$ equazioni, introducendo variabili ausiliarie per le derivate successive (matrice **compagna**).
- Le soluzioni $\mathbf x(t)$ si visualizzano come traiettorie nel **piano delle fasi**, tangenti in ogni punto al campo vettoriale $\mathbf x\mapsto A\mathbf x$.
