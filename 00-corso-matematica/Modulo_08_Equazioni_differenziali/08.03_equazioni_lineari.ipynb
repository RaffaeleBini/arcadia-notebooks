# 8.3 — Equazioni lineari

**Prerequisiti:** 8.2 Equazioni del primo ordine, 5.9 Teoremi fondamentali della derivazione.


## Perché ci interessa

L'equazione $y'=t-y$ (usata come esempio nel notebook 8.1) **non** è a variabili separabili: il membro destro $t-y$ non si fattorizza come prodotto di una funzione di $t$ e una di $y$. Ma ha un'altra struttura utile: è **lineare** in $y$ e $y'$ (nessun $y^2$, $y\cdot y'$, o $\sin y$). Le equazioni lineari del primo ordine formano una classe estremamente comune (compaiono in circuiti elettrici, miscele chimiche, dinamica delle popolazioni con immigrazione) e si risolvono con un trucco elegante: il **fattore integrante**, che trasforma il membro sinistro in una derivata di un prodotto, pronta per essere integrata direttamente.

## Teoria, passo dopo passo

### Forma standard

Un'equazione differenziale lineare del primo ordine si scrive nella forma standard:

$$y'(t) + p(t)\,y(t) = q(t)$$

(se $q(t)\equiv0$, si dice **omogenea**; altrimenti **non omogenea**). L'equazione $y'=t-y$ del notebook 8.1 si riscrive $y'+y=t$: forma standard con $p(t)=1$, $q(t)=t$.

### Il fattore integrante

L'idea: moltiplichiamo entrambi i membri per una funzione $\mu(t)$ (il **fattore integrante**), scelta apposta perché il membro sinistro diventi la derivata di un prodotto $\mu(t)y(t)$. Per la regola del prodotto (notebook 5.9): $\dfrac{d}{dt}(\mu y)=\mu'y+\mu y'$. Confrontando con $\mu y'+\mu p\,y$ (l'equazione moltiplicata per $\mu$), vogliamo $\mu'=\mu p$, cioè $\dfrac{\mu'}\mu=p(t)$ — un'equazione a variabili separabili (notebook 8.2) per $\mu$! Risolvendola:

$$\int\frac{d\mu}\mu = \int p(t)\,dt \quad\Longrightarrow\quad \ln\mu = \int p(t)\,dt \quad\Longrightarrow\quad \mu(t) = e^{\int p(t)\,dt}$$

(scegliendo la costante di integrazione uguale a zero, essendo sufficiente **un** fattore integrante qualsiasi, non tutti).

### Il procedimento completo

Con questo $\mu(t)$, l'equazione moltiplicata diventa $\dfrac{d}{dt}(\mu y)=\mu q(t)$ (per costruzione). Integrando entrambi i membri (teorema fondamentale del calcolo, notebook 5.16):

$$\mu(t)y(t) = \int\mu(t)q(t)\,dt + C \quad\Longrightarrow\quad y(t) = \frac1{\mu(t)}\left(\int\mu(t)q(t)\,dt+C\right)$$

### Esempio guidato: $y'+y=t$

$p(t)=1$, quindi $\mu(t)=e^{\int1\,dt}=e^t$. Moltiplicando: $e^ty'+e^ty=te^t$, cioè $\dfrac{d}{dt}(e^ty)=te^t$. Integrando il membro destro per parti (notebook 5.17, con $u=t,dv=e^tdt$): $\int te^t\,dt=te^t-e^t+C=(t-1)e^t+C$. Quindi $e^ty=(t-1)e^t+C$, cioè $y(t)=(t-1)+Ce^{-t}$ — esattamente la soluzione $y=t-1+Ce^{-t}$ usata (e verificata) nel notebook 8.1!

### Struttura della soluzione: omogenea + particolare

Notiamo che $y(t)=(t-1)+Ce^{-t}$ si decompone in due pezzi: $y_p(t)=t-1$ (una **soluzione particolare** dell'equazione non omogenea) più $y_h(t)=Ce^{-t}$ (la **soluzione generale dell'equazione omogenea associata** $y'+y=0$, risolvibile per separazione di variabili). Questa struttura — soluzione generale = soluzione particolare + soluzione omogenea generale — è un principio ricorrente in tutte le equazioni **lineari**, che ritroveremo (in forma più ricca) nel notebook 8.4 per le equazioni del secondo ordine.

````python
import sympy as sp

t = sp.symbols('t', real=True)
y_func = sp.Function('y')

# Risoluzione con il metodo del fattore integrante, passo per passo
p_t = 1
mu = sp.exp(sp.integrate(p_t, t))
print("Fattore integrante mu(t):", mu)

q_t = t
integranda = mu * q_t
integrale = sp.integrate(integranda, t)
print("Integrale di mu(t)*q(t):", integrale)

C = sp.symbols('C')
y_soluzione = (integrale + C) / mu
print("Soluzione generale:", sp.simplify(y_soluzione))

# Confronto con sp.dsolve
eq = sp.Eq(y_func(t).diff(t) + y_func(t), t)
print("\nConfronto con sp.dsolve:", sp.dsolve(eq, y_func(t)))
````

## Esercizio svolto e commentato

**Problema:** risolvi il problema di Cauchy $\begin{cases}y'+2ty=t\\y(0)=2\end{cases}$.

**Ragionamento:**

**Passo 1 — identifichiamo $p(t)=2t$ e $q(t)=t$** (forma standard già rispettata).

**Passo 2 — calcoliamo il fattore integrante**: $\mu(t)=e^{\int2t\,dt}=e^{t^2}$.

**Passo 3 — moltiplichiamo l'equazione per $\mu$**: $e^{t^2}y'+2te^{t^2}y=te^{t^2}$, cioè $\dfrac{d}{dt}\left(e^{t^2}y\right)=te^{t^2}$.

**Passo 4 — integriamo il membro destro**, con la sostituzione $u=t^2$, $du=2t\,dt$ (notebook 5.17): $\int te^{t^2}dt=\dfrac12\int e^u\,du=\dfrac12e^{t^2}+C_1$.

**Passo 5 — scriviamo l'equazione integrata**: $e^{t^2}y=\dfrac12e^{t^2}+C$, quindi $y(t)=\dfrac12+Ce^{-t^2}$.

**Passo 6 — imponiamo $y(0)=2$**: $2=\dfrac12+Ce^0=\dfrac12+C$, quindi $C=\dfrac32$.

**Conclusione:** $y(t)=\dfrac12+\dfrac32e^{-t^2}$.

**Perché il fattore integrante $e^{t^2}$ (calcolato al passo 2) "sa in anticipo" di dover produrre esattamente la derivata di un prodotto al passo 3, senza bisogno di verificarlo a mano ogni volta:** la costruzione del fattore integrante, derivata nella teoria, **garantisce per costruzione algebrica** che $\mu'=\mu p$, condizione che rende automaticamente $\mu y'+\mu p\,y=(\mu y)'$ (per la regola del prodotto, notebook 5.9) — non è una coincidenza fortunata specifica di questo esempio, ma una conseguenza diretta di come $\mu(t)=e^{\int p(t)dt}$ è stato costruito nella teoria: il metodo funziona sempre, per ogni equazione lineare del primo ordine in forma standard.

````python
import sympy as sp

t = sp.symbols('t', real=True)
y_func = sp.Function('y')

eq = sp.Eq(y_func(t).diff(t) + 2*t*y_func(t), t)
soluzione = sp.dsolve(eq, y_func(t), ics={y_func(0): 2})
print("Soluzione:", soluzione)
````

Imposta e risolvi l'equazione differenziale per $S(t)$, con $S(0)=5$.

## Riepilogo finale

- Un'equazione lineare del primo ordine $y'+p(t)y=q(t)$ si risolve con il **fattore integrante** $\mu(t)=e^{\int p(t)dt}$, che trasforma il membro sinistro in $\dfrac{d}{dt}(\mu y)$.
- Il procedimento riduce sempre l'equazione a un'integrazione diretta: $\mu y=\int\mu q\,dt+C$.
- La soluzione generale si decompone come **soluzione particolare + soluzione omogenea generale** — un principio che ritroveremo per le equazioni di ordine superiore.
- Il metodo funziona per costruzione, garantito dalla scelta algebrica di $\mu$: non serve verificarlo caso per caso.
