# 8.7 — Stabilità

**Prerequisiti:** 8.6 Matrici ed equazioni differenziali, 6.17 Autovalori.


## Perché ci interessa

Nel notebook 8.6 abbiamo risolto $\mathbf x'=A\mathbf x$ esplicitamente. Ma spesso interessa una domanda più qualitativa, e più immediata da rispondere: **cosa succede a lungo termine**? Le soluzioni tendono verso l'origine (l'equilibrio), se ne allontanano, o oscillano indefinitamente? Questa domanda — la **stabilità** dell'equilibrio — si risponde guardando **solo il segno** (o la parte reale) degli autovalori di $A$, senza nemmeno calcolare esplicitamente la soluzione. È un'informazione cruciale in ingegneria (un ponte, un aereo, un sistema di controllo devono essere stabili) e in biologia (un ecosistema torna all'equilibrio dopo una perturbazione?).

## Teoria, passo dopo passo

### Punto di equilibrio

Per il sistema $\mathbf x'=A\mathbf x$, il punto $\mathbf x=\mathbf 0$ è sempre un **equilibrio** (o punto fisso): se $\mathbf x(t_0)=\mathbf 0$ per qualche $t_0$, allora $\mathbf x(t)=\mathbf 0$ per ogni $t$ (la funzione costante nulla risolve banalmente l'equazione). La domanda di stabilità: le soluzioni **vicine** all'origine tendono ad avvicinarsi ulteriormente, ad allontanarsi, o restano a distanza limitata?

### Classificazione tramite gli autovalori (caso $2\times2$, reali distinti)

Per $A$ ($2\times2$) con autovalori reali distinti $\lambda_1,\lambda_2$, la soluzione generale è $\mathbf x(t)=C_1e^{\lambda_1t}\mathbf v_1+C_2e^{\lambda_2t}\mathbf v_2$ (notebook 8.6). Il comportamento per $t\to+\infty$ dipende dai segni:

- **entrambi negativi** ($\lambda_1,\lambda_2<0$): ogni termine $e^{\lambda_it}\to0$, quindi $\mathbf x(t)\to\mathbf 0$ per **ogni** condizione iniziale: l'origine è un **nodo stabile** (o pozzo), attrattivo;
- **entrambi positivi** ($\lambda_1,\lambda_2>0$): ogni termine cresce senza limite (tranne se $C_1=C_2=0$): l'origine è un **nodo instabile** (o sorgente), repulsivo;
- **segni opposti** ($\lambda_1<0<\lambda_2$): lungo la direzione di $\mathbf v_1$ le soluzioni si avvicinano all'origine, lungo $\mathbf v_2$ se ne allontanano: l'origine è una **sella**, instabile (quasi ogni traiettoria, tranne quelle esattamente lungo $\mathbf v_1$, finisce per allontanarsi).

### Autovalori complessi: spirali e centri

Se $\lambda=\alpha\pm i\beta$ (complessi coniugati, notebook 8.6), le soluzioni coinvolgono $e^{\alpha t}(\cos\beta t,\sin\beta t)$: il fattore $e^{\alpha t}$ controlla se le oscillazioni **crescono** o **si smorzano** in ampiezza.

- $\alpha<0$: **spirale stabile** (le traiettorie ruotano attorno all'origine, avvicinandosi);
- $\alpha>0$: **spirale instabile** (ruotano allontanandosi);
- $\alpha=0$ (autovalori immaginari puri): **centro**, le traiettorie sono **orbite chiuse** (come il cerchio dell'esercizio 4 del notebook 8.5) — né si avvicinano né si allontanano, un caso di stabilità "neutra" (marginale).

### Il criterio traccia-determinante

Ricordando dal notebook 6.17 che per una matrice $2\times2$, $\operatorname{tr}(A)=\lambda_1+\lambda_2$ e $\det A=\lambda_1\lambda_2$: si può classificare direttamente da questi due numeri, senza risolvere l'equazione caratteristica:

- $\det A<0$: autovalori reali di segno opposto $\to$ **sella**;
- $\det A>0$ e $\operatorname{tr}(A)<0$: entrambi gli autovalori (reali o parte reale, se complessi) negativi $\to$ **stabile** (nodo o spirale stabile);
- $\det A>0$ e $\operatorname{tr}(A)>0$: $\to$ **instabile**;
- $\det A>0$ e $\operatorname{tr}(A)=0$: $\to$ **centro** (caso limite).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def classifica(A):
    tr = A.trace()
    det = A.det()
    autoval = list(A.eigenvals().keys())
    return tr, det, autoval

# Esempi dei vari casi
matrici = {
    "Nodo stabile": sp.Matrix([[-1, 0], [0, -2]]),
    "Nodo instabile": sp.Matrix([[1, 0], [0, 2]]),
    "Sella": sp.Matrix([[1, 0], [0, -1]]),
    "Spirale stabile": sp.Matrix([[-0.5, -1], [1, -0.5]]),
    "Centro": sp.Matrix([[0, -1], [1, 0]]),
}

for nome, A in matrici.items():
    tr, det, autoval = classifica(A)
    print(f"{nome}: tr={tr}, det={det}, autovalori={autoval}")

# Visualizziamo il piano delle fasi per il caso "Spirale stabile"
A_spirale = np.array([[-0.5, -1], [1, -0.5]])
X, Y = np.meshgrid(np.linspace(-2,2,20), np.linspace(-2,2,20))
U = A_spirale[0,0]*X + A_spirale[0,1]*Y
V = A_spirale[1,0]*X + A_spirale[1,1]*Y

fig, ax = plt.subplots(figsize=(6,6))
ax.streamplot(X, Y, U, V, color='gray', density=1.2)
ax.set_title("Piano delle fasi: spirale stabile (autovalori con parte reale negativa)")
ax.set_aspect('equal')
plt.show()
````

## Esercizio svolto e commentato

**Problema:** classifica la stabilità dell'origine per il sistema con $A=\begin{pmatrix}-2&1\\1&-2\end{pmatrix}$.

**Ragionamento:**

**Passo 1 — calcoliamo traccia e determinante**: $\operatorname{tr}(A)=-2+(-2)=-4$. $\det A=(-2)(-2)-1\times1=4-1=3$.

**Passo 2 — applichiamo il criterio**: $\det A=3>0$ e $\operatorname{tr}(A)=-4<0$: siamo nel caso "**stabile**" (nodo o spirale stabile — per distinguerli serve il discriminante).

**Passo 3 — calcoliamo esplicitamente gli autovalori** per determinare se è un nodo (autovalori reali) o una spirale (complessi): equazione caratteristica $\lambda^2-\operatorname{tr}(A)\lambda+\det A=0$, cioè $\lambda^2+4\lambda+3=0$. Discriminante: $16-12=4>0$: **autovalori reali**, $\lambda=\dfrac{-4\pm2}2$, cioè $\lambda_1=-1,\lambda_2=-3$.

**Conclusione:** entrambi gli autovalori sono reali e negativi: l'origine è un **nodo stabile** (non una spirale, essendo gli autovalori reali, non complessi).

**Perché il criterio traccia-determinante (passo 2) è utile come primo controllo rapido, ma spesso serve comunque il passo 3 per una classificazione completa:** traccia e determinante da soli distinguono **stabile da instabile da sella**, ma non distinguono un **nodo** (autovalori reali) da una **spirale** (autovalori complessi) all'interno della categoria "stabile" (o "instabile") — quella distinzione richiede di controllare il segno del discriminante $\operatorname{tr}(A)^2-4\det A$ dell'equazione caratteristica: positivo per un nodo, negativo per una spirale. Il criterio rapido è quindi un primo filtro efficiente, da raffinare con il calcolo esplicito quando serve il quadro completo (utile, per esempio, per sapere se le traiettorie si avvicinano all'equilibrio "in linea retta" o "a spirale").

````python
import sympy as sp

A = sp.Matrix([[-2, 1], [1, -2]])
print("tr(A) =", A.trace(), " det(A) =", A.det())
print("Autovalori:", A.eigenvals())

discriminante = A.trace()**2 - 4*A.det()
print("Discriminante:", discriminante, "-> nodo (reali) se >0, spirale (complessi) se <0")
````

**1.** $\operatorname{tr}=4>0$, $\det=3>0$: **instabile** (nodo instabile, autovalori $3,1$ entrambi positivi, confermato direttamente).

**4.** $\operatorname{tr}(A)=0-1=-1$ (costante, non dipende da $k$). $\det A=0\times(-1)-1\times(-k)=k$. Per stabilità serve $\det A>0$ e $\operatorname{tr}(A)<0$: $\operatorname{tr}=-1<0$ sempre vero; serve $k>0$. Quindi: **stabile per ogni $k>0$**. Discriminante: $\operatorname{tr}^2-4\det=1-4k$. Se $k<\frac14$: discriminante positivo, **nodo stabile** (smorzamento "forte", nessuna oscillazione). Se $k>\frac14$: discriminante negativo, **spirale stabile** (oscillazioni smorzate) — coerente con l'intuizione fisica di uno smorzatore che, a seconda dei parametri, può far tornare il sistema all'equilibrio con o senza oscillazioni.

## Riepilogo finale

- L'origine, equilibrio di $\mathbf x'=A\mathbf x$, è classificata dal **segno degli autovalori** (o della loro parte reale, se complessi): entrambi negativi $\to$ stabile (nodo o spirale); entrambi positivi $\to$ instabile; segni opposti $\to$ sella; parte reale nulla $\to$ centro (stabilità neutra).
- Il **criterio traccia-determinante** ($\operatorname{tr}A<0,\det A>0\to$stabile; $\det A<0\to$sella; ecc.) permette una classificazione rapida senza risolvere l'equazione caratteristica.
- Il **discriminante** $\operatorname{tr}(A)^2-4\det A$ distingue nodi (positivo, autovalori reali) da spirali (negativo, autovalori complessi).
- Questa classificazione è centrale in ingegneria e nelle scienze applicate per determinare se un sistema torna spontaneamente all'equilibrio dopo una perturbazione.
