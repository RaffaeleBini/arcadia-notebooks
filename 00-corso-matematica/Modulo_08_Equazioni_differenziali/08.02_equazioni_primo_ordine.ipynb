## Perché ci interessa

Nel notebook 8.1 abbiamo visto cosa significa "risolvere" un'equazione differenziale, ma non abbiamo ancora un metodo sistematico per farlo. La classe più semplice e più comune di equazioni del primo ordine — le **equazioni a variabili separabili** — si risolve con una tecnica diretta: isolare tutta la dipendenza da $y$ da un lato e tutta la dipendenza da $t$ dall'altro, poi integrare entrambi i lati separatamente, riusando tutte le tecniche del notebook 5.17.

## Teoria, passo dopo passo

### Equazioni a variabili separabili

Un'equazione si dice a **variabili separabili** se si può scrivere nella forma:

$$y'(t) = g(t)\,h(y)$$

cioè il membro destro è un **prodotto** di una funzione della sola $t$ e una funzione della sola $y$. L'idea risolutiva: scriviamo $y'=\dfrac{dy}{dt}$ (notazione di Leibniz, che tratta $dy$ e $dt$ come quantità "separabili" — una manipolazione formale che si giustifica rigorosamente, ma che qui usiamo operativamente):

$$\frac{dy}{dt} = g(t)h(y) \quad\Longrightarrow\quad \frac{dy}{h(y)} = g(t)\,dt$$

(dividendo per $h(y)$, assumendo $h(y)\ne0$). Ora **integriamo entrambi i membri**, ciascuno rispetto alla propria variabile:

$$\int\frac{dy}{h(y)} = \int g(t)\,dt$$

Questo produce un'equazione (spesso implicita) che lega $y$ e $t$, con **una** costante di integrazione (le due costanti dei due integrali si combinano in una sola, potendo sempre portarle tutte da un lato).

### Perché il metodo è corretto (giustificazione rigorosa)

Il procedimento formale con $dy,dt$ nasconde, in realtà, un'applicazione della **regola della catena** (notebook 5.9) e del **teorema fondamentale del calcolo** (notebook 5.16). Sia $H(y)$ una primitiva di $1/h(y)$. Consideriamo $\dfrac{d}{dt}H(y(t))=H'(y(t))\,y'(t)=\dfrac{y'(t)}{h(y(t))}$ (regola della catena). Se $y(t)$ risolve $y'=g(t)h(y)$, allora $\dfrac{y'(t)}{h(y(t))}=g(t)$, quindi $\dfrac{d}{dt}H(y(t))=g(t)$. Integrando entrambi i membri rispetto a $t$ (teorema fondamentale del calcolo): $H(y(t))=\int g(t)\,dt+C$ — esattamente il risultato del metodo formale, ora giustificato con rigore.

### Esempio guidato: crescita esponenziale

Risolviamo $y'=ky$ (il modello di crescita/decadimento del notebook 8.1), con $k$ costante. Separando: $\dfrac{dy}y=k\,dt$. Integrando (per $y>0$, usando $\int\frac1ydy=\ln|y|$, notebook 5.17): $\ln|y|=kt+C_1$. Applicando l'esponenziale a entrambi i membri (invertendo il logaritmo, notebook 1.12): $|y|=e^{kt+C_1}=e^{C_1}e^{kt}$. Ponendo $C=\pm e^{C_1}$ (una costante che assorbe il segno e riscrive $e^{C_1}>0$ come un numero reale qualsiasi non nullo — e verificando poi che anche $C=0$, cioè $y\equiv0$, è una soluzione valida, persa nella divisione per $y$): $y(t)=Ce^{kt}$, la soluzione generale già usata nel notebook 8.1.

### Attenzione alle soluzioni "perse"

Dividere per $h(y)$ presuppone $h(y)\ne0$. Se $h(y_0)=0$ per qualche valore $y_0$, allora la funzione **costante** $y(t)\equiv y_0$ risolve banalmente $y'=0=g(t)h(y_0)$: è una soluzione **aggiuntiva** (detta soluzione **singolare** o **di equilibrio**), che il procedimento di separazione, dividendo per $h(y)$, "perde" e va sempre verificata separatamente.

````python
import sympy as sp

t, y = sp.symbols('t y', real=True)
C = sp.symbols('C')

# Esempio: y' = t*y^2 (equazione a variabili separabili)
# dy/y^2 = t dt  ->  -1/y = t^2/2 + C1  ->  y = -1/(t^2/2+C1)
y_func = sp.Function('y')
eq = sp.Eq(y_func(t).diff(t), t * y_func(t)**2)
soluzione = sp.dsolve(eq, y_func(t))
print("Soluzione di y' = t*y^2:", soluzione)

# Verifica manuale della separazione
integrale_y = sp.integrate(1/y**2, y)
integrale_t = sp.integrate(t, t)
print("\nIntegrale di dy/y^2:", integrale_y)
print("Integrale di t dt:", integrale_t)

# Soluzione di equilibrio (y=0 annulla h(y)=y^2)
print("\nVerifica: y=0 risolve y'=t*y^2?", sp.simplify(sp.diff(0, t) - t * 0**2) == 0)
````

## Esercizio svolto e commentato

**Problema:** risolvi il problema di Cauchy $\begin{cases}y'=\dfrac{t}y\\y(0)=3\end{cases}$ (per $y>0$).

**Ragionamento:**

**Passo 1 — riconosciamo la forma separabile**: $y'=\dfrac{t}y=t\times\dfrac1y$, con $g(t)=t$ e $h(y)=1/y$.

**Passo 2 — separiamo le variabili**: $\dfrac{dy}{1/y}=t\,dt$, cioè $y\,dy=t\,dt$.

**Passo 3 — integriamo entrambi i membri**: $\int y\,dy=\int t\,dt$, cioè $\dfrac{y^2}2=\dfrac{t^2}2+C_1$.

**Passo 4 — semplifichiamo**, moltiplicando per $2$ e rinominando $C=2C_1$: $y^2=t^2+C$.

**Passo 5 — imponiamo la condizione iniziale** $y(0)=3$: $3^2=0^2+C$, quindi $C=9$.

**Passo 6 — esplicitiamo $y$**, scegliendo il segno positivo della radice (coerente con la condizione iniziale $y(0)=3>0$): $y(t)=\sqrt{t^2+9}$.

**Conclusione:** la soluzione del problema di Cauchy è $y(t)=\sqrt{t^2+9}$, definita per **ogni** $t\in\mathbb R$ (l'argomento della radice è sempre positivo).

**Perché scegliere il segno corretto della radice (passo 6) è un passaggio delicato, da non automatizzare:** l'equazione $y^2=t^2+C$ ammette, a rigore, **due** soluzioni esplicite, $y=\sqrt{t^2+C}$ e $y=-\sqrt{t^2+C}$; solo verificando quale delle due soddisfa la condizione iniziale (qui, $y(0)=3>0$ esclude il ramo negativo) si individua la soluzione corretta del problema di Cauchy — un controllo che va sempre fatto esplicitamente, specialmente quando (come qui) l'equazione originale coinvolgeva $y$ al denominatore o sotto radice, situazioni in cui il segno non è automaticamente ovvio dalla sola equazione implicita.

````python
import sympy as sp

t = sp.symbols('t', real=True)
y_func = sp.Function('y')

eq = sp.Eq(y_func(t).diff(t), t / y_func(t))
soluzione = sp.dsolve(eq, y_func(t), ics={y_func(0): 3})
print("Soluzione del problema di Cauchy:", soluzione)
````

## Riepilogo finale

- Un'equazione a **variabili separabili** $y'=g(t)h(y)$ si risolve separando $\dfrac{dy}{h(y)}=g(t)\,dt$ e integrando entrambi i membri — un procedimento giustificabile rigorosamente con la regola della catena e il teorema fondamentale del calcolo.
- Va sempre controllato il **segno** nell'esplicitare $y$ da un'equazione implicita, usando la condizione iniziale.
- Se $h(y_0)=0$, la costante $y\equiv y_0$ è una **soluzione di equilibrio** aggiuntiva, "persa" dalla divisione — sempre da verificare separatamente.
- L'unicità della soluzione non è garantita se le ipotesi tecniche del teorema di esistenza e unicità falliscono (come per $h(y)=\sqrt{|y|}$ vicino a $y=0$).
