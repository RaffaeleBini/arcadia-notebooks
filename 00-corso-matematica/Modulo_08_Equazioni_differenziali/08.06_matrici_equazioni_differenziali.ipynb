# 8.6 — Matrici ed equazioni differenziali

**Prerequisiti:** 8.5 Sistemi di equazioni differenziali, 6.17 Autovalori, 6.18 Autovettori, 6.19 Diagonalizzazione.


## Perché ci interessa

Nel notebook 8.4, per risolvere $y'=ay$ (scalare), la soluzione era semplicemente $y=Ce^{at}$. Per il sistema $\mathbf x'=A\mathbf x$ (notebook 8.5), verrebbe naturale scrivere $\mathbf x(t)=e^{At}\mathbf x(0)$ — ma cosa significa "elevare $e$ a una **matrice**"? La risposta sta negli **autovalori e autovettori** di $A$ (Modulo 6): se conosciamo la diagonalizzazione di $A$, possiamo ricondurre il sistema accoppiato a **equazioni scalari indipendenti**, ciascuna risolvibile con il metodo del notebook 8.4.

## Teoria, passo dopo passo

### L'idea chiave: autovettori come soluzioni esponenziali

Cerchiamo soluzioni della forma $\mathbf x(t)=e^{\lambda t}\mathbf v$, con $\mathbf v$ vettore costante (l'analogo vettoriale del tentativo $y=e^{\lambda t}$ del notebook 8.4). Sostituendo in $\mathbf x'=A\mathbf x$: $\lambda e^{\lambda t}\mathbf v=Ae^{\lambda t}\mathbf v$, cioè (semplificando $e^{\lambda t}\ne0$):

$$A\mathbf v = \lambda\mathbf v$$

Riconosciamo esattamente l'equazione degli **autovalori e autovettori** (notebook 6.17-6.18)! Ogni coppia autovalore-autovettore $(\lambda,\mathbf v)$ di $A$ fornisce una soluzione $\mathbf x(t)=e^{\lambda t}\mathbf v$ del sistema.

### Caso diagonalizzabile con autovalori reali distinti

Se $A$ ($n\times n$) ha $n$ autovalori reali distinti $\lambda_1,\dots,\lambda_n$ con autovettori $\mathbf v_1,\dots,\mathbf v_n$ (automaticamente indipendenti, notebook 6.18), la **soluzione generale** del sistema è la combinazione lineare:

$$\mathbf x(t) = C_1e^{\lambda_1t}\mathbf v_1+C_2e^{\lambda_2t}\mathbf v_2+\cdots+C_ne^{\lambda_nt}\mathbf v_n$$

(si dimostra che ogni combinazione lineare di soluzioni resta soluzione, per la linearità del sistema — stesso principio del notebook 8.4 — e che, con $n$ autovettori indipendenti, questa famiglia a $n$ parametri esaurisce tutte le soluzioni). Le costanti $C_1,\dots,C_n$ si determinano dalla condizione iniziale $\mathbf x(0)$, risolvendo il sistema lineare $C_1\mathbf v_1+\cdots+C_n\mathbf v_n=\mathbf x(0)$ (notebook 6.11) — cioè trovando le coordinate di $\mathbf x(0)$ rispetto alla base di autovettori (notebook 6.8).

### Collegamento con la diagonalizzazione

Se $A=PDP^{-1}$ (notebook 6.19, con $P$ formata dagli autovettori e $D=\operatorname{diag}(\lambda_1,\dots,\lambda_n)$), il cambiamento di variabile $\mathbf x=P\mathbf y$ trasforma $\mathbf x'=A\mathbf x$ in:

$$P\mathbf y' = AP\mathbf y = PDP^{-1}P\mathbf y = PD\mathbf y \quad\Longrightarrow\quad \mathbf y' = D\mathbf y$$

(moltiplicando a sinistra per $P^{-1}$). Ma $\mathbf y'=D\mathbf y$, con $D$ diagonale, è un sistema di equazioni **disaccoppiate**: $y_i'=\lambda_iy_i$ per ogni $i$, ciascuna risolvibile indipendentemente (notebook 8.4) come $y_i(t)=C_ie^{\lambda_it}$! Tornando a $\mathbf x=P\mathbf y$, si ritrova esattamente la formula della soluzione generale sopra: la diagonalizzazione "disaccoppia" il sistema, trasformandolo in $n$ equazioni scalari indipendenti.

### Autovalori complessi: oscillazioni

Se $A$ ha autovalori complessi coniugati $\lambda=\alpha\pm i\beta$ (possibile anche per matrici $2\times2$ reali, come visto nel notebook 6.17), le soluzioni coinvolgono $e^{\alpha t}\cos(\beta t)$ e $e^{\alpha t}\sin(\beta t)$ — esattamente come nel Caso 3 del notebook 8.4, ora applicato alle componenti del vettore soluzione.

````python
import sympy as sp

t = sp.symbols('t', real=True)
A = sp.Matrix([[1, 2], [2, 1]])  # autovalori 3, -1 (dal notebook 6.18)

autovalori_autovettori = A.eigenvects()
print("Autovalori e autovettori di A:")
for lam, molt, vs in autovalori_autovettori:
    print(f"  lambda={lam}, v={vs[0].T}")

# Soluzione generale simbolica
C1, C2 = sp.symbols('C1 C2')
lam1, v1 = 3, sp.Matrix([1, 1])
lam2, v2 = -1, sp.Matrix([1, -1])

x_generale = C1 * sp.exp(lam1 * t) * v1 + C2 * sp.exp(lam2 * t) * v2
print("\nSoluzione generale x(t):", x_generale.T)

# Verifica: x'(t) = A*x(t) ?
x_prime = x_generale.diff(t)
Ax = A * x_generale
print("\nx'(t) =", sp.simplify(x_prime).T)
print("A*x(t) =", sp.simplify(Ax).T)
print("Coincidono?", sp.simplify(x_prime - Ax) == sp.zeros(2, 1))
````

## Esercizio svolto e commentato

**Problema:** risolvi il problema di Cauchy $\mathbf x'=A\mathbf x$ con $A=\begin{pmatrix}1&2\\2&1\end{pmatrix}$ (autovalori $3,-1$ con autovettori $(1,1),(1,-1)$, dal notebook 6.18) e $\mathbf x(0)=(4,2)$.

**Ragionamento:**

**Passo 1 — scriviamo la soluzione generale** usando gli autovalori/autovettori già noti: $\mathbf x(t)=C_1e^{3t}(1,1)+C_2e^{-t}(1,-1)$.

**Passo 2 — imponiamo la condizione iniziale** $\mathbf x(0)=(4,2)$: $C_1(1,1)+C_2(1,-1)=(4,2)$, cioè il sistema $\begin{cases}C_1+C_2=4\\C_1-C_2=2\end{cases}$.

**Passo 3 — risolviamo il sistema lineare** (notebook 6.11): sommando le due equazioni, $2C_1=6$, quindi $C_1=3$; sottraendo, $2C_2=2$, quindi $C_2=1$.

**Conclusione:** $\mathbf x(t)=3e^{3t}(1,1)+e^{-t}(1,-1)$, cioè componente per componente: $x(t)=3e^{3t}+e^{-t}$, $y(t)=3e^{3t}-e^{-t}$.

**Perché il passo 2 — trovare $C_1,C_2$ risolvendo un sistema lineare — è concettualmente lo stesso procedimento del notebook 6.8 (trovare le coordinate di un vettore rispetto a una base):** la condizione iniziale $\mathbf x(0)$ è un vettore fissato di $\mathbb R^2$, e $\{(1,1),(1,-1)\}$ è una base (essendo autovettori relativi ad autovalori distinti, quindi indipendenti, notebook 6.18); determinare $C_1,C_2$ significa esattamente trovare le **coordinate** di $\mathbf x(0)$ rispetto a questa base — un problema già risolto in astratto nel Modulo 6, qui applicato al caso concreto di un problema di Cauchy vettoriale.

````python
import sympy as sp

t = sp.symbols('t', real=True)
C1, C2 = sp.symbols('C1 C2')

v1, v2 = sp.Matrix([1, 1]), sp.Matrix([1, -1])
x0 = sp.Matrix([4, 2])

eq = C1 * v1 + C2 * v2 - x0
soluzione = sp.solve([eq[0], eq[1]], [C1, C2])
print("C1, C2:", soluzione)

x_t = soluzione[C1] * sp.exp(3*t) * v1 + soluzione[C2] * sp.exp(-t) * v2
print("\nx(t) =", x_t.T)
````

## Riepilogo finale

- Ogni coppia autovalore-autovettore $(\lambda,\mathbf v)$ di $A$ genera una soluzione $\mathbf x(t)=e^{\lambda t}\mathbf v$ del sistema $\mathbf x'=A\mathbf x$.
- Con autovalori reali distinti, la soluzione generale è $\mathbf x(t)=\sum_iC_ie^{\lambda_it}\mathbf v_i$; le costanti si trovano risolvendo un sistema lineare dalla condizione iniziale (coordinate rispetto alla base di autovettori, notebook 6.8).
- La **diagonalizzazione** $A=PDP^{-1}$ (notebook 6.19) "disaccoppia" il sistema in equazioni scalari indipendenti, ciascuna risolvibile col metodo del notebook 8.4.
- Autovalori complessi producono soluzioni oscillanti, con la stessa struttura del Caso 3 del notebook 8.4.
