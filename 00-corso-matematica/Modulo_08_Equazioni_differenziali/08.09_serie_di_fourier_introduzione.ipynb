# 8.9 — Serie di Fourier (introduzione)

**Prerequisiti:** 5.20 Serie di potenze (rigorose), 6.20 Prodotti scalari, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Nel notebook 5.20 abbiamo visto che molte funzioni si possono scrivere come somma infinita di **potenze** $x^n$ (serie di Taylor). Esiste un'idea alternativa, altrettanto potente: scrivere una funzione **periodica** come somma infinita di **seni e coseni**. Perché seni e coseni? Perché sono le funzioni periodiche più semplici possibili, e — come vedremo — formano un sistema di "assi coordinati" reciprocamente ortogonali (nel senso del prodotto scalare tra funzioni, notebook 6.20), esattamente come i versori $e_1,\dots,e_n$ in $\mathbb R^n$. Questa idea, dovuta a Fourier (che la sviluppò proprio studiando la propagazione del calore, un'equazione differenziale), è alla base di applicazioni che vanno dall'elaborazione audio alla compressione delle immagini.

## Teoria, passo dopo passo

### Funzioni periodiche e la serie di Fourier

Una funzione $f$ è **periodica** di periodo $2\pi$ se $f(x+2\pi)=f(x)$ per ogni $x$. La **serie di Fourier** di $f$ è la (candidata) rappresentazione:

$$f(x) \sim \frac{a_0}2 + \sum_{n=1}^\infty\left(a_n\cos(nx)+b_n\sin(nx)\right)$$

dove i **coefficienti di Fourier** $a_n,b_n$ si calcolano con formule integrali specifiche (sotto). Il simbolo $\sim$ (invece di $=$) segnala che, in generale, servono condizioni tecniche perché la somma della serie coincida effettivamente con $f(x)$ punto per punto — una sottigliezza che il Modulo 15 approfondirà; qui lavoriamo assumendo che $f$ sia "abbastanza buona" (per esempio continua a tratti) perché l'uguaglianza valga.

### Ortogonalità di seni e coseni

Il prodotto scalare tra funzioni introdotto nel notebook 6.20, $\langle f,g\rangle=\int f(x)g(x)\,dx$, applicato sull'intervallo $[-\pi,\pi]$, rivela un fatto notevole: le funzioni $1,\cos x,\sin x,\cos2x,\sin2x,\dots$ sono **ortogonali** tra loro (notebook 6.21), cioè $\int_{-\pi}^\pi\cos(nx)\cos(mx)\,dx=0$ per $n\ne m$ (e analogamente per le altre combinazioni), verificabile con le formule di prostaferesi (identità trigonometriche, notebook 2.10).

### Le formule dei coefficienti

Esattamente come nel notebook 6.21 le coordinate rispetto a una base ortogonale si trovavano con la formula $\lambda_i=\langle u,v_i\rangle/\langle v_i,v_i\rangle$, qui i coefficienti di Fourier si trovano "proiettando" $f$ su ciascuna funzione della base ortogonale $\{1,\cos(nx),\sin(nx)\}$:

$$a_n = \frac1\pi\int_{-\pi}^\pi f(x)\cos(nx)\,dx \qquad b_n = \frac1\pi\int_{-\pi}^\pi f(x)\sin(nx)\,dx$$

(il fattore $1/\pi$ compensa $\langle\cos(nx),\cos(nx)\rangle=\pi$, calcolabile esplicitamente). Queste formule sono **l'analogo continuo** delle coordinate rispetto a una base ortogonale (notebook 6.21), con la somma finita sostituita da un integrale, e lo spazio vettoriale $\mathbb R^n$ sostituito dallo spazio (infinito-dimensionale) delle funzioni.

### Simmetria: funzioni pari e dispari

Se $f$ è **pari** ($f(-x)=f(x)$, notebook 3.8), tutti i $b_n=0$ (essendo $\sin(nx)$ dispari, l'integrale di un prodotto pari-dispari su un intervallo simmetrico è nullo, per lo stesso principio di simmetria del notebook 7.15): la serie contiene solo coseni. Se $f$ è **dispari**, tutti gli $a_n=0$: solo seni. Riconoscere la simmetria di $f$ dimezza subito il lavoro.

````python
import sympy as sp

x, n = sp.symbols('x n', real=True)

# Serie di Fourier dell'onda quadra: f(x) = 1 per 0<x<pi, -1 per -pi<x<0 (dispari)
f = sp.Piecewise((-1, x < 0), (1, x >= 0))

# Essendo dispari, calcoliamo solo b_n
n_val = sp.symbols('n', positive=True, integer=True)
bn = sp.Rational(1, 1) / sp.pi * sp.integrate(f * sp.sin(n_val * x), (x, -sp.pi, sp.pi))
bn = sp.simplify(bn)
print("b_n per l'onda quadra:", bn)

# Valutiamo i primi coefficienti
for i in range(1, 6):
    print(f"b_{i} =", bn.subs(n_val, i))

# Verifica ortogonalità: integrale di cos(2x)*cos(3x) su [-pi,pi]
ortho = sp.integrate(sp.cos(2*x) * sp.cos(3*x), (x, -sp.pi, sp.pi))
print("\nIntegrale di cos(2x)*cos(3x) su [-pi,pi]:", ortho, "(0, coerente con l'ortogonalità)")

norma_cos = sp.integrate(sp.cos(2*x)**2, (x, -sp.pi, sp.pi))
print("Integrale di cos(2x)^2 su [-pi,pi]:", norma_cos, "(coerente col fattore pi nelle formule)")
````

## Esercizio svolto e commentato

**Problema:** calcola i primi coefficienti di Fourier di $f(x)=x$ su $[-\pi,\pi]$ (estesa periodicamente).

**Ragionamento:**

**Passo 1 — riconosciamo che $f(x)=x$ è una funzione **dispari**** ($f(-x)=-x=-f(x)$): per il principio di simmetria della teoria, tutti gli $a_n=0$ (incluso $a_0$), senza bisogno di calcolare nessun integrale con i coseni.

**Passo 2 — calcoliamo $b_n$**: $b_n=\dfrac1\pi\displaystyle\int_{-\pi}^\pi x\sin(nx)\,dx$. Poiché $x\sin(nx)$ è **pari** (prodotto di due funzioni dispari), l'integrale su $[-\pi,\pi]$ è il doppio dell'integrale su $[0,\pi]$: $b_n=\dfrac2\pi\displaystyle\int_0^\pi x\sin(nx)\,dx$.

**Passo 3 — integriamo per parti** (notebook 5.17), con $u=x,\,dv=\sin(nx)dx$, quindi $du=dx,\,v=-\dfrac{\cos(nx)}n$: $\displaystyle\int_0^\pi x\sin(nx)dx=\left[-\dfrac{x\cos(nx)}n\right]_0^\pi+\dfrac1n\displaystyle\int_0^\pi\cos(nx)\,dx$.

**Passo 4 — valutiamo il primo termine**: $-\dfrac{\pi\cos(n\pi)}n-0=-\dfrac{\pi(-1)^n}n$ (usando $\cos(n\pi)=(-1)^n$, notebook 2.9). Il secondo termine: $\dfrac1n\left[\dfrac{\sin(nx)}n\right]_0^\pi=\dfrac1{n^2}(\sin(n\pi)-0)=0$ (essendo $\sin(n\pi)=0$ per $n$ intero).

**Passo 5 — componiamo**: $\displaystyle\int_0^\pi x\sin(nx)dx=-\dfrac{\pi(-1)^n}n$. Quindi $b_n=\dfrac2\pi\times\left(-\dfrac{\pi(-1)^n}n\right)=-\dfrac{2(-1)^n}n=\dfrac{2(-1)^{n+1}}n$.

**Conclusione:** $f(x)=x\sim\displaystyle\sum_{n=1}^\infty\dfrac{2(-1)^{n+1}}n\sin(nx)=2\left(\sin x-\dfrac{\sin2x}2+\dfrac{\sin3x}3-\cdots\right)$.

**Perché riconoscere la parità/disparità (passi 1-2) prima di lanciarsi nel calcolo può dimezzare (letteralmente) il lavoro:** oltre ad annullare a priori metà dei coefficienti (passo 1), la disparità di $f$ permette anche di raddoppiare-e-dimezzare l'intervallo di integrazione (passo 2, integrando su $[0,\pi]$ invece di $[-\pi,\pi]$ e moltiplicando per 2) — un espediente che semplifica sistematicamente ogni calcolo di coefficienti di Fourier per funzioni con simmetria nota, esattamente come le simmetrie semplificavano gli integrali doppi nel notebook 7.15.

````python
import sympy as sp

x, n = sp.symbols('x n', positive=True, integer=True)

bn = sp.Rational(2, 1) / sp.pi * sp.integrate(x * sp.sin(n * x), (x, 0, sp.pi))
bn = sp.simplify(bn)
print("b_n per f(x)=x:", bn)

for i in range(1, 5):
    print(f"b_{i} =", bn.subs(n, i))
````

## Riepilogo finale

- La **serie di Fourier** rappresenta una funzione periodica come somma di seni e coseni: $f(x)\sim\frac{a_0}2+\sum(a_n\cos nx+b_n\sin nx)$.
- Seni e coseni sono **ortogonali** tra loro (rispetto al prodotto scalare tra funzioni, notebook 6.20-6.21): i coefficienti si trovano "proiettando" $f$ su ciascuna funzione di base, con formule integrali analoghe a quelle del notebook 6.21.
- Funzioni **pari** hanno solo coefficienti $a_n$ (coseni); funzioni **dispari** hanno solo $b_n$ (seni) — un fatto che dimezza il lavoro di calcolo.
- Le serie di Fourier collegano l'algebra lineare astratta (basi ortogonali, Modulo 6) alle applicazioni concrete di analisi armonica.
