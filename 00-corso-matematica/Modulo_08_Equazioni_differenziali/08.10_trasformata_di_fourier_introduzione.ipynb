# 8.10 — Trasformata di Fourier (introduzione)

**Prerequisiti:** 8.9 Serie di Fourier (introduzione), 5.18 Integrali impropri.


## Perché ci interessa

La serie di Fourier del notebook 8.9 funziona per funzioni **periodiche**. Ma molti segnali — un singolo impulso, un suono che inizia e finisce, una funzione che decade a zero — non sono periodici. La **trasformata di Fourier** estende l'idea a questo caso, sostituendo la somma discreta su frequenze intere $n$ con un **integrale continuo** su tutte le frequenze possibili — pensando, intuitivamente, a una funzione non periodica come il "limite" di una funzione periodica con periodo che tende a infinito.

## Teoria, passo dopo passo

### Dall'intuizione: periodo che tende a infinito

Nella serie di Fourier (notebook 8.9), le frequenze $n$ (per periodo $2\pi$) sono discrete: $1,2,3,\dots$. Se il periodo $T$ aumenta, le frequenze permesse $2\pi n/T$ si infittiscono sempre di più; nel limite $T\to\infty$ (funzione non più periodica), le frequenze diventano un **continuo**, e la somma discreta $\sum_n$ si trasforma naturalmente in un **integrale** su una variabile di frequenza continua $\omega$.

### Definizione

La **trasformata di Fourier** di $f:\mathbb R\to\mathbb R$ (o $\mathbb C$) è:

$$\hat f(\omega) = \int_{-\infty}^{+\infty}f(x)e^{-i\omega x}\,dx$$

(un integrale improprio, notebook 5.18; qui usiamo la notazione complessa $e^{-i\omega x}=\cos(\omega x)-i\sin(\omega x)$, formalizzata nel Modulo 9, che unifica seno e coseno in un'unica espressione compatta — l'analogo continuo dei coefficienti $a_n,b_n$ del notebook 8.9). La funzione $\hat f(\omega)$ dice "quanto" della frequenza $\omega$ è presente in $f$: il suo modulo $|\hat f(\omega)|$ si chiama **spettro** di $f$.

### La trasformata inversa

Come una serie di Fourier ricostruisce $f$ sommando i suoi termini pesati, la **trasformata inversa** ricostruisce $f$ integrando $\hat f$ su tutte le frequenze:

$$f(x) = \frac1{2\pi}\int_{-\infty}^{+\infty}\hat f(\omega)e^{i\omega x}\,d\omega$$

Questa coppia di trasformazioni (diretta e inversa) permette di passare avanti e indietro tra la rappresentazione "nel tempo" (o nello spazio) $f(x)$ e quella "in frequenza" $\hat f(\omega)$ — due descrizioni equivalenti della stessa informazione, utili per scopi diversi (per esempio, filtrare un segnale è spesso più facile lavorando su $\hat f$).

### Esempio: la trasformata di una funzione gaussiana

Un caso notevole (e calcolabile esplicitamente, riusando l'integrale gaussiano del notebook 7.15): $f(x)=e^{-x^2}$. Si dimostra (calcolo che omettiamo nei dettagli, ma verifichiamo numericamente) che $\hat f(\omega)=\sqrt\pi\,e^{-\omega^2/4}$ — **ancora** una gaussiana! Questa proprietà (la gaussiana è "autoduale" rispetto alla trasformata di Fourier, a meno di riscalamenti) è centrale in probabilità (distribuzione normale) e in fisica quantistica (principio di indeterminazione di Heisenberg, che è essenzialmente un'affermazione sulla trasformata di Fourier).

### Perché è collegata alle equazioni differenziali

Un fatto cruciale: la trasformata di Fourier trasforma la **derivazione** in una semplice **moltiplicazione**: si dimostra che $\widehat{f'}(\omega)=i\omega\hat f(\omega)$ (integrando per parti, notebook 5.17, nella definizione della trasformata). Questo significa che un'equazione differenziale nello spazio $x$ diventa un'equazione **algebrica** nello spazio delle frequenze $\omega$ — un'idea che sfrutteremo sistematicamente, nella sua forma "gemella" più pratica per problemi con condizioni iniziali, con la trasformata di Laplace nel prossimo notebook.

````python
import sympy as sp

x, omega = sp.symbols('x omega', real=True)

# Trasformata di Fourier della gaussiana f(x) = exp(-x^2)
f = sp.exp(-x**2)
f_hat = sp.integrate(f * sp.exp(-sp.I * omega * x), (x, -sp.oo, sp.oo))
f_hat = sp.simplify(f_hat)
print("Trasformata di Fourier di exp(-x^2):", f_hat)

# Verifica numerica della proprietà "derivata -> moltiplicazione per i*omega"
# usando una funzione più semplice: f(x) = exp(-x^2), f'(x) = -2x*exp(-x^2)
f_prime = sp.diff(f, x)
f_prime_hat = sp.integrate(f_prime * sp.exp(-sp.I * omega * x), (x, -sp.oo, sp.oo))
f_prime_hat = sp.simplify(f_prime_hat)
print("\nTrasformata di f'(x):", f_prime_hat)
print("i*omega*f_hat(omega):", sp.simplify(sp.I * omega * f_hat))
print("Coincidono?", sp.simplify(f_prime_hat - sp.I * omega * f_hat) == 0)
````

## Esercizio svolto e commentato

**Problema:** calcola la trasformata di Fourier della funzione "rettangolo" $f(x)=\begin{cases}1&|x|\le1\\0&|x|>1\end{cases}$ (un impulso di durata finita, un segnale tipico non periodico).

**Ragionamento:**

**Passo 1 — impostiamo l'integrale**, che si riduce a un intervallo finito grazie al supporto compatto di $f$: $\hat f(\omega)=\displaystyle\int_{-1}^1e^{-i\omega x}\,dx$ (fuori da $[-1,1]$, $f=0$, quindi non contribuisce).

**Passo 2 — integriamo direttamente** (una primitiva elementare dell'esponenziale complesso, trattato come nel caso reale, notebook 5.9): $\displaystyle\int e^{-i\omega x}dx=\dfrac{e^{-i\omega x}}{-i\omega}$ (per $\omega\ne0$).

**Passo 3 — valutiamo tra $-1$ e $1$**: $\hat f(\omega)=\dfrac{e^{-i\omega}-e^{i\omega}}{-i\omega}=\dfrac{e^{i\omega}-e^{-i\omega}}{i\omega}$.

**Passo 4 — riconosciamo la formula di Eulero** (anticipata nella teoria): $e^{i\omega}-e^{-i\omega}=2i\sin\omega$ (sottraendo $\cos\omega-i\sin\omega$ da $\cos\omega+i\sin\omega$, i termini coseno si cancellano). Sostituendo: $\hat f(\omega)=\dfrac{2i\sin\omega}{i\omega}=\dfrac{2\sin\omega}\omega$.

**Passo 5 — controlliamo il caso $\omega=0$** (escluso al passo 2): $\hat f(0)=\int_{-1}^1 1\,dx=2$, e $\lim_{\omega\to0}\dfrac{2\sin\omega}\omega=2\times1=2$ (limite notevole, notebook 5.5): coerente, la formula si estende con continuità anche a $\omega=0$.

**Conclusione:** $\hat f(\omega)=\dfrac{2\sin\omega}\omega$ — una funzione reale (nonostante l'integrale coinvolgesse numeri complessi), perché $f$ è **pari**: un fatto generale, analogo a quanto visto per le serie di Fourier nel notebook 8.9 (funzioni pari hanno rappresentazioni "senza componente immaginaria").

**Perché il risultato $\frac{2\sin\omega}\omega$ (una funzione chiamata "sinc", onnipresente nell'elaborazione dei segnali) ha senso intuitivamente:** lo spettro di un impulso "corto e netto" nello spazio (qui, un rettangolo di larghezza $2$) è "largo e diffuso" in frequenza (il sinc decade lentamente, oscillando); questo è un caso particolare del **principio di indeterminazione** anticipato nella teoria: più una funzione è concentrata in $x$, più la sua trasformata è dispersa in $\omega$, e viceversa (la gaussiana, che bilancia perfettamente concentrazione in entrambi i domini, è il caso estremo di questo compromesso).

````python
import sympy as sp

x, omega = sp.symbols('x omega', real=True)
f_rect = sp.Piecewise((1, sp.Abs(x) <= 1), (0, True))

f_hat = sp.integrate(f_rect * sp.exp(-sp.I * omega * x), (x, -sp.oo, sp.oo))
f_hat = sp.simplify(f_hat)
print("Trasformata di Fourier del rettangolo:", f_hat)
````

**1.** Espandendo $e^{-i\omega x}=\cos(\omega x)-i\sin(\omega x)$: $\hat f(\omega)=\int f(x)\cos(\omega x)dx-i\int f(x)\sin(\omega x)dx$. **3.** $\widehat{f''}(\omega)=i\omega\widehat{f'}(\omega)=i\omega\times i\omega\hat f(\omega)=-\omega^2\hat f(\omega)$ (applicando la proprietà due volte).

**4.** Trasformando entrambi i membri: $-\widehat{f''}(\omega)+\hat f(\omega)=\hat g(\omega)$. ## Riepilogo finale

- La **trasformata di Fourier** $\hat f(\omega)=\int_{-\infty}^{+\infty}f(x)e^{-i\omega x}dx$ estende la serie di Fourier a funzioni non periodiche, sostituendo la somma discreta con un integrale continuo sulle frequenze.
- La **trasformata inversa** ricostruisce $f$ da $\hat f$, esattamente come la serie di Fourier ricostruiva $f$ dai suoi coefficienti.
- La derivazione diventa **moltiplicazione per $i\omega$** nello spazio delle frequenze: un fatto che trasforma equazioni differenziali in equazioni algebriche.
- Funzioni concentrate in $x$ hanno trasformate disperse in $\omega$ (e viceversa): un compromesso noto come principio di indeterminazione.
