## Perché ci interessa

Finora abbiamo sempre lavorato con funzioni **date esplicitamente**: $f(x)=x^2$, $f(x)=\sin x$, ecc. Ma moltissimi fenomeni naturali non forniscono la funzione direttamente: forniscono una **relazione tra la funzione e le sue derivate**. La legge di raffreddamento di Newton dice che la velocità con cui un oggetto si raffredda è proporzionale alla differenza di temperatura con l'ambiente: $T'(t)=-k(T(t)-T_{amb})$. La crescita di una popolazione con risorse illimitate soddisfa $P'(t)=kP(t)$. In entrambi i casi, non conosciamo $T$ o $P$ direttamente: conosciamo solo **come cambiano**. Un'**equazione differenziale** è precisamente questo: un'equazione che coinvolge una funzione incognita e le sue derivate. Risolverla significa trovare la funzione stessa.

## Teoria, passo dopo passo

### Definizione ed esempio più semplice

Un'**equazione differenziale ordinaria** (EDO) è un'equazione della forma $F(t,y,y',y'',\dots,y^{(n)})=0$, dove $y=y(t)$ è la funzione incognita. L'**ordine** dell'equazione è l'ordine massimo di derivata che compare. L'esempio più semplice possibile: $y'(t)=f(t)$, la cui soluzione è semplicemente una primitiva, $y(t)=\int f(t)\,dt$ (teorema fondamentale del calcolo, notebook 5.16) — un'equazione differenziale del primo ordine che abbiamo già, in un certo senso, risolto per tutto il Modulo 5, senza chiamarla così.

### Soluzione generale e particolare

Consideriamo $y'(t)=2t$. Le soluzioni sono $y(t)=t^2+C$, per **qualsiasi** costante $C$ (essendo la derivata di una costante nulla, notebook 5.8): infinite funzioni diverse risolvono la stessa equazione! Questa famiglia si chiama **soluzione generale**. Se imponiamo una condizione aggiuntiva, come $y(0)=3$ (una **condizione iniziale**), determiniamo univocamente $C$: $y(0)=0+C=3\Rightarrow C=3$, dando la **soluzione particolare** $y(t)=t^2+3$.

### Il problema di Cauchy

Un'equazione differenziale del primo ordine insieme a una condizione iniziale, $\begin{cases}y'=f(t,y)\\y(t_0)=y_0\end{cases}$, si chiama **problema di Cauchy** (o problema ai valori iniziali). Sotto ipotesi ragionevoli su $f$ (continuità, e una condizione tecnica detta di Lipschitz, che non approfondiamo qui), il **teorema di esistenza e unicità** garantisce che esiste **esattamente una** soluzione — non zero, non infinite: proprio come una condizione iniziale identifica un unico punto su una retta tangente (notebook 5.8), qui identifica un'unica curva soluzione tra tutte quelle della famiglia generale.

### Campo di direzioni: visualizzare senza risolvere

Per un'equazione $y'=f(t,y)$, anche senza risolverla esplicitamente, possiamo **visualizzarne il comportamento**: in ogni punto $(t,y)$ del piano, l'equazione ci dice la pendenza $f(t,y)$ che la soluzione deve avere lì. Disegnando piccoli segmenti con quella pendenza in una griglia di punti, si ottiene il **campo di direzioni** (o campo di pendenze): le soluzioni sono curve che, in ogni punto, sono tangenti al segmento locale — un modo per "vedere" la forma delle soluzioni prima ancora di calcolarle.

### Perché le EDO sono ovunque

Equazioni come $y'=ky$ (crescita/decadimento esponenziale, notebook 1.11), $y''+\omega^2y=0$ (oscillatore armonico, che risolveremo nel notebook 8.4), o sistemi accoppiati (notebook 8.5) descrivono fenomeni in fisica, biologia, economia, ingegneria: qualunque situazione in cui il **tasso di cambiamento** di una grandezza dipende dalla grandezza stessa (o da altre grandezze correlate).

````python
import numpy as np
import matplotlib.pyplot as plt

# Campo di direzioni per y' = t - y
def f(t, y):
    return t - y

T = np.linspace(-2, 2, 20)
Y = np.linspace(-2, 2, 20)
Tg, Yg = np.meshgrid(T, Y)
slopes = f(Tg, Yg)

# Normalizziamo per disegnare segmenti di lunghezza uniforme
norm = np.sqrt(1 + slopes**2)
dx = 1 / norm
dy = slopes / norm

fig, ax = plt.subplots(figsize=(7, 7))
ax.quiver(Tg, Yg, dx, dy, angles='xy', headwidth=0, headlength=0, headaxislength=0, color='gray', alpha=0.6)

# Sovrapponiamo alcune soluzioni particolari (y = t - 1 + C*e^(-t), verificabile per sostituzione)
for C in [-2, -1, 0, 1, 2]:
    t_vals = np.linspace(-2, 2, 100)
    y_vals = t_vals - 1 + C * np.exp(-t_vals)
    ax.plot(t_vals, y_vals, linewidth=1.5)

ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.set_xlabel('t')
ax.set_ylabel('y')
ax.set_title("Campo di direzioni per y' = t - y, con alcune soluzioni")
plt.show()

# Verifica che y = t - 1 + C*e^(-t) risolve y' = t - y
import sympy as sp
t, C = sp.symbols('t C')
y = t - 1 + C * sp.exp(-t)
y_prime = sp.diff(y, t)
print("y' =", y_prime)
print("t - y =", sp.expand(t - y))
print("Coincidono?", sp.simplify(y_prime - (t - y)) == 0)
````

## Esercizio svolto e commentato

**Problema:** verifica che $y(t)=Ce^{2t}$ è soluzione generale di $y'=2y$, poi trova la soluzione particolare con $y(0)=5$.

**Ragionamento:**

**Passo 1 — verifichiamo che $y(t)=Ce^{2t}$ risolve l'equazione**, calcolando $y'(t)$: $y'(t)=C\times2e^{2t}=2Ce^{2t}$ (derivata dell'esponenziale, notebook 5.9).

**Passo 2 — confrontiamo con $2y(t)$**: $2y(t)=2Ce^{2t}$. Coincide esattamente con $y'(t)$ trovato al passo 1: l'equazione $y'=2y$ è soddisfatta, per **qualsiasi** valore di $C$.

**Passo 3 — imponiamo la condizione iniziale** $y(0)=5$: $y(0)=Ce^{2\times0}=Ce^0=C\times1=C$. Quindi $C=5$.

**Conclusione:** la soluzione particolare è $y(t)=5e^{2t}$.

**Perché la costante $C$ nella soluzione generale di un'equazione del primo ordine gioca lo stesso ruolo della costante di integrazione del notebook 5.16:** risolvere $y'=2y$ equivale, concettualmente, a "integrare" l'equazione — e come ogni integrazione lascia indeterminata una costante additiva (perché la derivata di una costante è zero), qui l'equazione differenziale lascia indeterminato un fattore moltiplicativo $C$ (perché, come vedremo nel notebook 8.2, l'equazione $y'=2y$ si risolve separando le variabili e integrando, un procedimento che introduce esattamente una costante di integrazione). Una singola condizione iniziale — un solo numero — basta a fissarla univocamente, esattamente come per le equazioni del tipo $y'=f(t)$.

````python
import sympy as sp

t, C = sp.symbols('t C')
y = C * sp.exp(2 * t)

y_prime = sp.diff(y, t)
print("y'(t) =", y_prime)
print("2*y(t) =", sp.expand(2 * y))
print("Verificata l'equazione y'=2y?", sp.simplify(y_prime - 2*y) == 0)

C_valore = sp.solve(sp.Eq(y.subs(t, 0), 5), C)
print("\nValore di C per y(0)=5:", C_valore)
````

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** $y'=C\cos t-D\sin t$, $y''=-C\sin t-D\cos t$. $y''+y=(-C\sin t-D\cos t)+(C\sin t+D\cos t)=0$. ✓ Verificata per ogni $C,D$.

**2.** L'ordine è determinato dalla derivata di ordine più alto presente: $y'''$ compare, quindi l'equazione è di **terzo ordine** (il termine $y^2$, pur non lineare, non cambia l'ordine, che riguarda solo il numero di derivazioni, non le potenze della funzione stessa).

**3.** $y(0)=Ce^0=C=-2$. Soluzione particolare: $y(t)=-2e^{-3t}$.

**4.** $y'=\dfrac{d}{dt}\left(\dfrac1{1-t}\right)=\dfrac1{(1-t)^2}$ (regola della catena, notebook 5.9, su $(1-t)^{-1}$). $y^2=\left(\dfrac1{1-t}\right)^2=\dfrac1{(1-t)^2}$: coincide con $y'$. ✓ $y(0)=\dfrac1{1-0}=1$. ✓ Per $t\to1^-$, $1-t\to0^+$, quindi $y(t)=\dfrac1{1-t}\to+\infty$: la soluzione "esplode" (ha un'asintoto verticale) in $t=1$, e non è definita oltre. Questo fenomeno, chiamato **blow-up in tempo finito**, è tipico delle equazioni **non lineari** (qui il termine $y^2$): a differenza delle equazioni lineari come $y'=ky$ (le cui soluzioni esponenziali sono definite per ogni $t$), le equazioni non lineari possono avere soluzioni che esistono solo su un intervallo limitato di tempo.

## Riepilogo finale

- Un'**equazione differenziale** lega una funzione incognita alle sue derivate; l'**ordine** è quello della derivata più alta presente.
- La **soluzione generale** contiene costanti arbitrarie (una per ogni ordine di integrazione necessario); una **condizione iniziale** le fissa, dando la **soluzione particolare** (**problema di Cauchy**).
- Il **campo di direzioni** permette di visualizzare qualitativamente le soluzioni senza risolvere esplicitamente l'equazione.
- Le equazioni **non lineari** possono avere soluzioni che "esplodono" in tempo finito, un fenomeno assente nelle equazioni lineari a coefficienti costanti.
