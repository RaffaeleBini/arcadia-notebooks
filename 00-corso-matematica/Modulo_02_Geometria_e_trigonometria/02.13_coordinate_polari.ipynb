# 2.13 — Coordinate polari

**Prerequisiti:** 2.1 Piano cartesiano, 2.3 Distanza tra punti, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Nel notebook 2.1 abbiamo descritto un punto del piano con una coppia $(x,y)$: quanto ci si sposta orizzontalmente e verticalmente dall'origine. Ma per certe curve — in particolare quelle con una forte simmetria rispetto a un punto, come una spirale, o un fenomeno che dipende naturalmente da una **distanza** e una **direzione** (un radar, un'orbita) — c'è un sistema di coordinate più naturale: descrivere un punto tramite **quanto è lontano** dall'origine e **in che direzione**. Sono le **coordinate polari**, e useremo direttamente seno e coseno (notebook 2.9) per collegarle a quelle cartesiane.

## Teoria, passo dopo passo

### Definizione

Un punto $P$ del piano (diverso dall'origine) si può descrivere con due numeri:

- $\rho$ (rho), la **distanza** di $P$ dall'origine ($\rho\ge0$) — calcolabile con la formula del notebook 2.3.
- $\theta$, l'**angolo** che il segmento $OP$ forma con il semiasse positivo delle $x$, misurato come nel notebook 2.8.

La coppia $(\rho,\theta)$ sono le **coordinate polari** di $P$. L'origine stessa ha $\rho=0$ (e $\theta$ non è definito in modo univoco: qualunque angolo "va bene" per il punto a distanza zero).

### Conversione da polari a cartesiane

Se pensiamo a $P$ come al punto sulla circonferenza di raggio $\rho$ (invece che $1$, come nella circonferenza goniometrica, notebook 2.8-2.9) individuato dall'angolo $\theta$, le coordinate cartesiane si ottengono scalando le definizioni di seno e coseno per il raggio $\rho$:

$$x=\rho\cos\theta \qquad y=\rho\sin\theta$$

(quando $\rho=1$, ritroviamo esattamente le definizioni del notebook 2.9: coordinate polari e circonferenza goniometrica sono strettamente imparentate).

### Conversione da cartesiane a polari

Dato $(x,y)$, troviamo $\rho$ con la formula della distanza dall'origine (notebook 2.3, caso particolare con un punto nell'origine):

$$\rho = \sqrt{x^2+y^2}$$

Per $\theta$, dalla definizione di tangente (notebook 2.9): $\tan\theta=\frac{y}{x}$ (per $x\ne0$), quindi $\theta=\arctan\left(\frac{y}{x}\right)$ — **con attenzione**: la funzione arcotangente (che formalizzeremo nel Modulo 3, parlando di funzioni inverse) restituisce sempre un angolo tra $-\frac{\pi}{2}$ e $\frac{\pi}{2}$, quindi bisogna **correggere** il risultato di $\pi$ se il punto $(x,y)$ si trova nel II o III quadrante (dove $x<0$), esattamente come abbiamo dovuto scegliere il segno corretto quando risolvevamo per il quadrante nel notebook 2.9.

### Curve in coordinate polari

Alcune curve hanno un'equazione molto più semplice in coordinate polari che in cartesiane. La circonferenza di raggio $R$ centrata nell'origine (notebook 2.4, equazione cartesiana $x^2+y^2=R^2$) diventa semplicemente:

$$\rho = R$$

(ogni punto a distanza $R$ dall'origine, indipendentemente dall'angolo — esattamente la definizione di circonferenza). Una curva più caratteristica delle coordinate polari è la **spirale di Archimede**, $\rho=a\theta$ (con $a$ costante): la distanza dall'origine cresce proporzionalmente all'angolo, producendo una spirale che si allontana man mano che gira — una curva la cui equazione cartesiana sarebbe molto più complicata da scrivere.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

rho, theta = sp.symbols('rho theta', positive=True)

# Conversione polari -> cartesiane
def polari_a_cartesiane(rho_val, theta_val):
    return rho_val * np.cos(theta_val), rho_val * np.sin(theta_val)

x, y = polari_a_cartesiane(5, np.pi/3)
print(f"Punto con rho=5, theta=pi/3: cartesiane ({x:.3f}, {y:.3f})")

# Conversione cartesiane -> polari
def cartesiane_a_polari(x_val, y_val):
    rho_val = np.sqrt(x_val**2 + y_val**2)
    theta_val = np.arctan2(y_val, x_val)  # arctan2 gestisce automaticamente il quadrante corretto
    return rho_val, theta_val

rho_val, theta_val = cartesiane_a_polari(-3, 4)
print(f"\nPunto (-3,4): polari (rho={rho_val:.3f}, theta={theta_val:.3f} rad = {np.degrees(theta_val):.1f}°)")

# Grafico della spirale di Archimede
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw={'projection': 'polar'})
theta_vals = np.linspace(0, 4*np.pi, 500)
a = 0.5
rho_vals = a * theta_vals
ax.plot(theta_vals, rho_vals)
ax.set_title("Spirale di Archimede: ρ = 0.5·θ")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** converti il punto cartesiano $(-1,-1)$ in coordinate polari.

**Ragionamento:**

**Passo 1 — calcoliamo $\rho$** con la formula della distanza (notebook 2.3):

$$\rho = \sqrt{(-1)^2+(-1)^2} = \sqrt{2}$$

**Passo 2 — troviamo un primo candidato per $\theta$** usando $\tan\theta=\frac{y}{x}=\frac{-1}{-1}=1$. L'angolo con tangente $1$ nel primo quadrante (dalla tabella dei valori notevoli, notebook 2.9) è $\theta_0=\frac{\pi}{4}$.

**Passo 3 — determiniamo il quadrante corretto.** Il punto $(-1,-1)$ ha $x<0$ e $y<0$: è nel **III quadrante** (notebook 2.1). Ma $\theta_0=\frac{\pi}{4}$ è un angolo del I quadrante: dobbiamo correggerlo. Poiché la tangente ha periodo $\pi$ (notebook 2.11), l'angolo del III quadrante con la stessa tangente di $\frac{\pi}{4}$ è:

$$\theta = \theta_0+\pi = \frac{\pi}{4}+\pi=\frac{5\pi}{4}$$

**Verifica**: convertiamo $(\rho,\theta)=\left(\sqrt2,\frac{5\pi}{4}\right)$ di nuovo in cartesiane: $x=\sqrt2\cos\frac{5\pi}{4}=\sqrt2\times\left(-\frac{\sqrt2}{2}\right)=-1$ ✓; $y=\sqrt2\sin\frac{5\pi}{4}=\sqrt2\times\left(-\frac{\sqrt2}{2}\right)=-1$ ✓.

**Perché non possiamo fermarci al primo valore di $\theta$ trovato dall'arcotangente:** come osservato nella teoria, l'arcotangente restituisce sempre un valore tra $-\frac{\pi}{2}$ e $\frac{\pi}{2}$ (I o IV quadrante), indipendentemente dal quadrante reale del punto. Guardare **entrambi** i segni di $x$ e $y$ (non solo il loro rapporto $\frac{y}{x}$, che perde l'informazione sul segno di ciascuno separatamente) è indispensabile per scegliere l'angolo giusto — lo stesso tipo di attenzione richiesta nel notebook 2.9 per scegliere il segno corretto di $\cos\theta$ dato $\sin\theta$.

````python
import numpy as np

x_val, y_val = -1, -1
rho_val = np.sqrt(x_val**2 + y_val**2)
theta_val = np.arctan2(y_val, x_val)  # gestisce automaticamente il quadrante
print(f"rho = {rho_val:.4f}")
print(f"theta = {theta_val:.4f} rad = {np.degrees(theta_val):.1f}°")

# Verifica riconvertendo
x_verifica = rho_val * np.cos(theta_val)
y_verifica = rho_val * np.sin(theta_val)
print(f"Verifica: ({x_verifica:.4f}, {y_verifica:.4f})")
````

## Riepilogo finale

- Le coordinate polari $(\rho,\theta)$ descrivono un punto tramite la distanza dall'origine e l'angolo rispetto al semiasse positivo delle $x$.
- Conversione polari→cartesiane: $x=\rho\cos\theta$, $y=\rho\sin\theta$. Conversione cartesiane→polari: $\rho=\sqrt{x^2+y^2}$, $\theta=\arctan(y/x)$ **corretto in base al quadrante** di $(x,y)$.
- Curve con simmetria rispetto all'origine (come la circonferenza centrata nell'origine, o la spirale) hanno equazioni molto più semplici in coordinate polari.
