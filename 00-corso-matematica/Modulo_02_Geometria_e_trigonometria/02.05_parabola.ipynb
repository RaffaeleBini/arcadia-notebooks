# 2.5 — Parabola

**Prerequisiti:** 1.6 Equazioni di secondo grado, 2.3 Distanza tra punti.


## Perché ci interessa

Nel notebook 1.6 abbiamo risolto equazioni come $x^2-5x+6=0$, trovando i valori di $x$ che le soddisfano. Ma cosa succede se disegniamo, punto per punto, tutti i valori $(x,y)$ con $y=x^2-5x+6$ (non solo dove $y=0$, ma per **ogni** $x$)? Otteniamo una curva a forma di "U", la **parabola** — e le soluzioni dell'equazione originale sono esattamente i punti dove questa curva incrocia l'asse $x$.

Questo notebook completa il cerchio: l'equazione di secondo grado, che finora era solo un calcolo algebrico, diventa anche un disegno, con un significato geometrico per ogni sua parte (il discriminante, i coefficienti...).

## Teoria, passo dopo passo

### Definizione come luogo geometrico

Una parabola è l'insieme dei punti equidistanti da un punto fisso, detto **fuoco**, e da una retta fissa, detta **direttrice**. Questa definizione (analoga nello spirito a quella della circonferenza, notebook 2.4, ma con una retta al posto di un centro) porta, con qualche calcolo tramite la formula della distanza (notebook 2.3), a un'equazione della forma:

$$y=ax^2+bx+c \qquad (a\ne0)$$

Non svolgiamo qui tutti i passaggi (richiedono di trattare il fuoco e la direttrice esplicitamente), ma partiamo direttamente da questa equazione, che è esattamente il polinomio di secondo grado del notebook 1.6, letto ora come funzione $y=f(x)$.

### Il vertice e l'asse di simmetria

Il punto più basso (se $a>0$) o più alto (se $a<0$) della parabola si chiama **vertice**. Per trovarlo, completiamo il quadrato (stessa tecnica del notebook 1.6 e del notebook 2.4):

$$y=ax^2+bx+c = a\left(x+\frac{b}{2a}\right)^2 + c-\frac{b^2}{4a}$$

Il termine $\left(x+\frac{b}{2a}\right)^2$ è sempre $\ge0$, ed è $0$ esattamente quando $x=-\frac{b}{2a}$: questo è dove $y$ raggiunge il suo valore minimo (se $a>0$) o massimo (se $a<0$). Il vertice è quindi:

$$V=\left(-\frac{b}{2a},\ -\frac{\Delta}{4a}\right)$$

dove $\Delta=b^2-4ac$ è lo stesso discriminante del notebook 1.6 (verificabile sostituendo $x=-\frac{b}{2a}$ nell'equazione e semplificando). La retta verticale $x=-\frac{b}{2a}$ è l'**asse di simmetria**: la parabola è simmetrica rispetto a essa, perché $y(x)$ dipende da $\left(x+\frac{b}{2a}\right)^2$, e il quadrato "non vede" il segno.

### Concavità

Il segno di $a$ determina la forma:

- $a>0$: la parabola ha la concavità **verso l'alto** (una "U"), il vertice è un **minimo**.
- $a<0$: concavità **verso il basso** (una "∩"), il vertice è un **massimo**.

### Il discriminante, ora con un significato geometrico

Le intersezioni con l'asse $x$ (dove $y=0$) sono le soluzioni di $ax^2+bx+c=0$, esattamente il problema del notebook 1.6! Il discriminante $\Delta$ ci dice ora quante volte la parabola **tocca** l'asse $x$:

- $\Delta>0$: la parabola **attraversa** l'asse $x$ in due punti distinti.
- $\Delta=0$: la parabola è **tangente** all'asse $x$ nel vertice (un solo punto di contatto).
- $\Delta<0$: la parabola **non tocca mai** l'asse $x$ (resta tutta sopra, se $a>0$, o tutta sotto, se $a<0$).

Quello che nel notebook 1.6 era un fatto puramente algebrico ("quante soluzioni reali") ha ora un'interpretazione visiva diretta.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')

def analizza_parabola(a, b, c):
    delta = b**2 - 4*a*c
    x_vertice = -b / (2*a)
    y_vertice = -delta / (4*a)
    return delta, (x_vertice, y_vertice)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
casi = [(1, -4, 3, "delta>0: due intersezioni"),
        (1, -4, 4, "delta=0: tangente"),
        (1, 0, 4, "delta<0: nessuna intersezione")]

for ax, (a, b, c, titolo) in zip(axes, casi):
    delta, (xv, yv) = analizza_parabola(a, b, c)
    x_vals = np.linspace(xv - 4, xv + 4, 200)
    y_vals = a*x_vals**2 + b*x_vals + c
    ax.plot(x_vals, y_vals)
    ax.axhline(y=0, color='gray', linewidth=0.8)
    ax.plot(xv, yv, 'ro', label=f'vertice ({xv},{yv})')
    ax.set_title(f"{titolo}\ny={a}x²+{b}x+{c}, Δ={delta}")
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Verifica simbolica della formula del vertice
a, b, c = sp.symbols('a b c')
espressione = a*x**2 + b*x + c
forma_completata = sp.expand(a*(x + b/(2*a))**2 + c - b**2/(4*a))
print("Coincide con l'espressione originale?", sp.simplify(espressione - forma_completata) == 0)
````

## Esercizio svolto e commentato

**Problema:** studia la parabola $y=-2x^2+8x-6$: trova vertice, concavità, e intersezioni con l'asse $x$.

**Ragionamento:**

**Passo 1 — identifichiamo i coefficienti**: $a=-2$, $b=8$, $c=-6$.

**Passo 2 — concavità**: $a=-2<0$, quindi la parabola ha concavità **verso il basso**, il vertice è un massimo.

**Passo 3 — calcoliamo il discriminante**:

$$\Delta = b^2-4ac = 8^2-4(-2)(-6) = 64-48=16$$

$\Delta=16>0$: la parabola attraversa l'asse $x$ in due punti.

**Passo 4 — troviamo il vertice**:

$$x_V = -\frac{b}{2a} = -\frac{8}{2(-2)} = -\frac{8}{-4}=2$$

$$y_V = -\frac{\Delta}{4a} = -\frac{16}{4(-2)} = -\frac{16}{-8}=2$$

Vertice: $V=(2,2)$ (un massimo, coerente con la concavità verso il basso).

**Passo 5 — troviamo le intersezioni con l'asse $x$**, risolvendo $-2x^2+8x-6=0$ (o equivalentemente $2x^2-8x+6=0$, dividendo per $-1$, notebook 0.5) con la formula risolutiva (notebook 1.6):

$$x=\frac{8\pm\sqrt{16}}{4}=\frac{8\pm4}{4} \quad\Rightarrow\quad x_1=3,\ x_2=1$$

Intersezioni: $(1,0)$ e $(3,0)$.

**Verifica di coerenza**: l'asse di simmetria $x=2$ dovrebbe essere esattamente a metà tra le due intersezioni $x_1=3$ e $x_2=1$: $\frac{3+1}{2}=2$. ✓ Coerente con la simmetria della parabola rispetto al proprio asse.

**Perché la media delle due intersezioni con l'asse $x$ coincide sempre con l'ascissa del vertice:** è una conseguenza diretta della simmetria vista nella teoria — la parabola è simmetrica rispetto alla retta $x=-\frac{b}{2a}$, quindi se attraversa l'asse $x$ in due punti, questi devono essere equidistanti dall'asse di simmetria, e la loro media cade esattamente lì.

````python
import sympy as sp

x = sp.symbols('x')
a, b, c = -2, 8, -6

delta = b**2 - 4*a*c
x_vertice = sp.Rational(-b, 2*a)
y_vertice = sp.Rational(-delta, 4*a)
print(f"Delta={delta}, Vertice=({x_vertice},{y_vertice})")

intersezioni = sp.solve(sp.Eq(a*x**2 + b*x + c, 0), x)
print("Intersezioni con l'asse x:", intersezioni)
print("Media delle intersezioni:", sum(intersezioni)/2, " (deve coincidere con x del vertice)")
````

## Riepilogo finale

- La parabola $y=ax^2+bx+c$ ($a\ne0$) è il grafico dell'equazione di secondo grado (notebook 1.6), letta come funzione.
- Il **vertice** è $V=\left(-\frac{b}{2a},-\frac{\Delta}{4a}\right)$, trovato completando il quadrato; è un minimo se $a>0$, un massimo se $a<0$.
- Il discriminante $\Delta$ ha ora un significato visivo: $\Delta>0$ due intersezioni con l'asse $x$, $\Delta=0$ tangenza nel vertice, $\Delta<0$ nessuna intersezione.
- La parabola è simmetrica rispetto alla retta verticale $x=-\frac{b}{2a}$: se ci sono due intersezioni con l'asse $x$, la loro media coincide sempre con l'ascissa del vertice.
