## Perché ci interessa

Nel notebook 0.19 abbiamo definito le **coppie ordinate** $(a,b)$ e il prodotto cartesiano $A\times B$: strumenti puramente algebrici, fatti di numeri e simboli. Ma c'è un modo per **vederli**: se disegniamo due rette perpendicolari e associamo a ogni coppia $(x,y)$ di numeri reali un punto del piano, otteniamo una rappresentazione visiva di $\mathbb{R}\times\mathbb{R}$.

Questa idea — nata da Cartesio (da cui il nome "cartesiano") — è il ponte tra algebra e geometria: da qui in avanti, ogni equazione con due incognite $x,y$ diventerà anche un disegno, e ogni figura geometrica diventerà anche un'equazione. Useremo questo doppio linguaggio in tutto il resto del corso.

## Teoria, passo dopo passo

### Costruzione del piano cartesiano

Disegniamo due rette numeriche (notebook 0.4) perpendicolari tra loro, che si incontrano in un punto comune chiamato **origine**, indicato $O$. La retta orizzontale si chiama **asse delle ascisse** (o asse $x$), quella verticale **asse delle ordinate** (asse $y$). Insieme, formano il **piano cartesiano**.

Ogni punto $P$ del piano è individuato da una **coppia ordinata** $(x_P,y_P)$ (notebook 0.19): $x_P$ è la sua proiezione sull'asse $x$ (l'**ascissa**), $y_P$ è la sua proiezione sull'asse $y$ (l'**ordinata**). L'ordine conta (proprio come nella definizione di coppia ordinata): il punto $(2,3)$ è diverso dal punto $(3,2)$.

**Corrispondenza biunivoca**: a ogni punto del piano corrisponde esattamente una coppia $(x,y)\in\mathbb{R}\times\mathbb{R}$, e viceversa a ogni coppia corrisponde esattamente un punto. Questo è ciò che rende il piano cartesiano un "disegno" fedele di $\mathbb{R}\times\mathbb{R}$: non perdiamo né aggiungiamo informazione passando dall'uno all'altro.

### I quadranti

I due assi dividono il piano in quattro regioni, dette **quadranti**, numerati in senso antiorario a partire da quello in alto a destra:

- **I quadrante**: $x>0$ e $y>0$.
- **II quadrante**: $x<0$ e $y>0$.
- **III quadrante**: $x<0$ e $y<0$.
- **IV quadrante**: $x>0$ e $y<0$.

I punti con $x=0$ stanno sull'asse $y$; i punti con $y=0$ stanno sull'asse $x$; l'origine $(0,0)$ sta su entrambi.

### Da un'equazione a un disegno: un primo assaggio

Consideriamo l'insieme di tutti i punti $(x,y)$ che soddisfano $y=x$ (una condizione, non un singolo punto). Se proviamo alcuni valori — $(0,0)$, $(1,1)$, $(2,2)$, $(-1,-1)$ — notiamo che stanno tutti allineati, su una retta che passa per l'origine a $45°$. Questo è il primo esempio di ciò che faremo sistematicamente da qui in avanti: un'equazione in $x,y$ descrive un **luogo geometrico**, l'insieme di tutti i punti che la soddisfano — un caso specifico dell'idea di "proprietà caratteristica" già vista per gli insiemi (notebook 0.12), applicata qui a un sottoinsieme del piano.

````python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 6))

# Assi cartesiani
ax.axhline(y=0, color='black', linewidth=1)
ax.axvline(x=0, color='black', linewidth=1)

# Alcuni punti nei quattro quadranti
punti = {"(2,3) - I quadrante": (2, 3), "(-2,3) - II quadrante": (-2, 3),
         "(-2,-3) - III quadrante": (-2, -3), "(2,-3) - IV quadrante": (2, -3)}
for etichetta, (x, y) in punti.items():
    ax.plot(x, y, 'o', markersize=8)
    ax.annotate(etichetta, (x, y), textcoords="offset points", xytext=(10, 5), fontsize=8)

# La retta y=x, come primo esempio di "equazione -> disegno"
x_vals = np.linspace(-4, 4, 10)
ax.plot(x_vals, x_vals, '--', color='gray', label='y = x')

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title("Il piano cartesiano: quadranti, punti, e un primo luogo geometrico")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dati i punti $A=(3,-2)$, $B=(-1,-2)$, $C=(-1,4)$, $D=(3,4)$, stabilisci in quale quadrante si trova ciascuno (o su quale asse), e riconosci quale figura geometrica formano.

**Ragionamento:**

**Classifichiamo ogni punto** guardando i segni delle coordinate (definizione dei quadranti vista nella teoria):

- $A=(3,-2)$: $x=3>0$, $y=-2<0$ → **IV quadrante**.
- $B=(-1,-2)$: $x=-1<0$, $y=-2<0$ → **III quadrante**.
- $C=(-1,4)$: $x=-1<0$, $y=4>0$ → **II quadrante**.
- $D=(3,4)$: $x=3>0$, $y=4>0$ → **I quadrante**.

**Riconosciamo la figura.** Osserviamo che $A$ e $D$ hanno la stessa ascissa ($x=3$): stanno sulla stessa retta verticale. Allo stesso modo $B$ e $C$ hanno $x=-1$: altra retta verticale. Analogamente, $A$ e $B$ hanno la stessa ordinata ($y=-2$): stessa retta orizzontale; $C$ e $D$ hanno $y=4$: altra retta orizzontale.

Quattro lati, a due a due paralleli agli assi, di conseguenza a due a due paralleli tra loro e perpendicolari agli altri due: i quattro punti sono i vertici di un **rettangolo**.

**Calcoliamo le dimensioni** (anticipando il notebook 2.3 sulla distanza): il lato orizzontale ha lunghezza $|3-(-1)|=4$ (differenza delle ascisse, notebook 0.8 valore assoluto come distanza); il lato verticale ha lunghezza $|4-(-2)|=6$.

**Perché riconoscere "stessa ascissa" o "stessa ordinata" è la chiave per identificare rapidamente lati orizzontali/verticali:** due punti con la stessa ordinata stanno necessariamente sulla stessa retta orizzontale (nessun altro valore di $y$ compare), e viceversa per l'ascissa. Questo criterio, puramente algebrico, corrisponde esattamente all'osservazione geometrica "i punti sono allineati orizzontalmente/verticalmente" — un altro esempio del ponte algebra-geometria che stiamo costruendo.

````python
import matplotlib.pyplot as plt

punti = {"A": (3, -2), "B": (-1, -2), "C": (-1, 4), "D": (3, 4)}

fig, ax = plt.subplots(figsize=(6, 6))
ax.axhline(y=0, color='black', linewidth=1)
ax.axvline(x=0, color='black', linewidth=1)

# Disegniamo il rettangolo collegando i punti in ordine
ordine = ["A", "B", "C", "D", "A"]
xs = [punti[p][0] for p in ordine]
ys = [punti[p][1] for p in ordine]
ax.plot(xs, ys, 'o-', color='tab:blue')
for nome, (x, y) in punti.items():
    ax.annotate(f"{nome}({x},{y})", (x, y), textcoords="offset points", xytext=(8, 8))

ax.set_xlim(-4, 5)
ax.set_ylim(-4, 6)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title("Il rettangolo ABCD")
plt.show()

# Dimensioni
larghezza = abs(3 - (-1))
altezza = abs(4 - (-2))
print(f"Larghezza (lato orizzontale): {larghezza}")
print(f"Altezza (lato verticale): {altezza}")
````

## Riepilogo finale

- Il piano cartesiano associa a ogni coppia ordinata $(x,y)\in\mathbb{R}\times\mathbb{R}$ (notebook 0.19) un unico punto del piano, e viceversa: è la rappresentazione visiva del prodotto cartesiano $\mathbb{R}\times\mathbb{R}$.
- Gli assi $x$ (ascisse) e $y$ (ordinate) si incontrano nell'**origine** $(0,0)$ e dividono il piano in quattro **quadranti**, distinti dai segni di $x$ e $y$.
- Un'equazione in due incognite $x,y$ descrive, in generale, un **luogo geometrico**: l'insieme dei punti del piano che la soddisfano.
- Punti con la stessa ascissa stanno su una retta verticale; punti con la stessa ordinata stanno su una retta orizzontale — un criterio algebrico immediato per riconoscere allineamenti.
