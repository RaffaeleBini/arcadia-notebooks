# 2.10 — Identità trigonometriche

**Prerequisiti:** 2.9 Seno, coseno, tangente, 1.2 Prodotti notevoli.


## Perché ci interessa

Conosciamo $\sin$ e $\cos$ per pochi angoli notevoli (notebook 2.9): $0,\frac{\pi}{6},\frac{\pi}{4},\frac{\pi}{3},\frac{\pi}{2}$. Ma come calcoliamo $\sin(75°)$? Notiamo che $75°=45°+30°$: se avessimo una formula per $\sin(\alpha+\beta)$ in funzione di $\sin\alpha,\cos\alpha,\sin\beta,\cos\beta$, potremmo costruire moltissimi altri valori esatti a partire da quelli noti.

Queste **formule di addizione**, insieme a poche altre identità derivate, sono gli strumenti algebrici della trigonometria: ci permetteranno, nel prossimo notebook, di risolvere equazioni trigonometriche più complesse di $\sin\theta=k$.

## Teoria, passo dopo passo

### Le formule di addizione (senza dimostrazione completa)

Non dimostriamo qui la derivazione geometrica completa (richiede un confronto tra distanze sulla circonferenza goniometrica, notebook 2.3 e 2.9), ma enunciamo le formule, che userai costantemente:

$$\sin(\alpha+\beta) = \sin\alpha\cos\beta+\cos\alpha\sin\beta$$
$$\cos(\alpha+\beta) = \cos\alpha\cos\beta-\sin\alpha\sin\beta$$

Sostituendo $\beta$ con $-\beta$, e ricordando che $\cos(-\beta)=\cos\beta$ (il coseno è una funzione **pari**: $P$ e il suo simmetrico rispetto all'asse $x$ hanno la stessa ascissa) e $\sin(-\beta)=-\sin\beta$ (il seno è **dispari**: ordinate opposte), otteniamo le formule di sottrazione:

$$\sin(\alpha-\beta) = \sin\alpha\cos\beta-\cos\alpha\sin\beta$$
$$\cos(\alpha-\beta) = \cos\alpha\cos\beta+\sin\alpha\sin\beta$$

### Formule di duplicazione

Ponendo $\beta=\alpha$ nelle formule di addizione, otteniamo le **formule di duplicazione**:

$$\sin(2\alpha) = 2\sin\alpha\cos\alpha$$

$$\cos(2\alpha) = \cos^2\alpha-\sin^2\alpha$$

La formula per $\cos(2\alpha)$ si può riscrivere in altre due forme equivalenti, usando l'identità fondamentale $\cos^2\alpha+\sin^2\alpha=1$ (notebook 2.9) per eliminare una delle due funzioni:

$$\cos(2\alpha) = 2\cos^2\alpha-1 = 1-2\sin^2\alpha$$

(sostituendo $\sin^2\alpha=1-\cos^2\alpha$ nella prima forma si ottiene la seconda, e viceversa — un utile esempio di come una singola identità fondamentale generi diverse forme equivalenti, tutte utili in contesti diversi).

### Formula per la tangente della somma

Dividendo la formula di $\sin(\alpha+\beta)$ per quella di $\cos(\alpha+\beta)$ (definizione di tangente, notebook 2.9), e poi dividendo numeratore e denominatore per $\cos\alpha\cos\beta$:

$$\tan(\alpha+\beta) = \frac{\tan\alpha+\tan\beta}{1-\tan\alpha\tan\beta}$$

### Perché queste formule sono davvero "identità"

Chiamiamo queste relazioni **identità** (non semplicemente "formule") perché valgono per **ogni** valore di $\alpha$ e $\beta$, non solo per alcuni: sono equazioni sempre vere (esattamente come le identità algebriche del notebook 1.5, dove un'equazione poteva risultare "indeterminata", vera per ogni $x$). Verificarle su singoli valori (come faremo nella cella di codice) non le dimostra in generale, ma è un buon modo per controllarne la plausibilità.

````python
import sympy as sp

alpha, beta = sp.symbols('alpha beta', real=True)

# Verifica simbolica delle formule di addizione (SymPy le conosce già internamente)
formula_seno_somma = sp.sin(alpha)*sp.cos(beta) + sp.cos(alpha)*sp.sin(beta)
print("sin(a+b) - formula:", sp.simplify(sp.sin(alpha+beta) - formula_seno_somma))

formula_coseno_somma = sp.cos(alpha)*sp.cos(beta) - sp.sin(alpha)*sp.sin(beta)
print("cos(a+b) - formula:", sp.simplify(sp.cos(alpha+beta) - formula_coseno_somma))

# Verifica delle formule di duplicazione
print("\nsin(2a) - 2 sin(a)cos(a):", sp.simplify(sp.sin(2*alpha) - 2*sp.sin(alpha)*sp.cos(alpha)))
print("cos(2a) - (cos²a - sin²a):", sp.simplify(sp.cos(2*alpha) - (sp.cos(alpha)**2 - sp.sin(alpha)**2)))

# Verifica numerica su un valore specifico, come controllo di plausibilità
import numpy as np
a_val, b_val = np.pi/5, np.pi/7
print(f"\nControllo numerico con a={a_val:.4f}, b={b_val:.4f}:")
print("sin(a+b) diretto:", np.sin(a_val + b_val))
print("formula:", np.sin(a_val)*np.cos(b_val) + np.cos(a_val)*np.sin(b_val))
````

## Esercizio svolto e commentato

**Problema:** calcola il valore esatto di $\sin(75°)$, usando $75°=45°+30°$.

**Ragionamento:**

**Passo 1 — applichiamo la formula di addizione per il seno** con $\alpha=45°$, $\beta=30°$:

$$\sin(75°)=\sin(45°+30°)=\sin45°\cos30°+\cos45°\sin30°$$

**Passo 2 — sostituiamo i valori notevoli** (notebook 2.9): $\sin45°=\frac{\sqrt2}{2}$, $\cos30°=\frac{\sqrt3}{2}$, $\cos45°=\frac{\sqrt2}{2}$, $\sin30°=\frac12$:

$$\sin(75°) = \frac{\sqrt2}{2}\cdot\frac{\sqrt3}{2} + \frac{\sqrt2}{2}\cdot\frac12$$

**Passo 3 — svolgiamo i prodotti** (notebook 0.6, radici):

$$= \frac{\sqrt6}{4} + \frac{\sqrt2}{4}$$

**Passo 4 — raccogliamo il denominatore comune** (già uguale, $4$):

$$\sin(75°) = \frac{\sqrt6+\sqrt2}{4}$$

**Verifica numerica**: $\frac{\sqrt6+\sqrt2}{4}\approx\frac{2.449+1.414}{4}\approx\frac{3.863}{4}\approx0.966$, che coincide con il valore di $\sin(75°)$ calcolato direttamente.

**Perché scegliere $75°=45°+30°$ e non un'altra scomposizione (per esempio $75°=90°-15°$):** entrambe le scomposizioni funzionerebbero, ma $45°+30°$ usa solo angoli il cui seno e coseno **conosciamo già esattamente** (tabella del notebook 2.9). Scegliere sempre una scomposizione in termini di angoli noti è la strategia chiave per applicare le formule di addizione in modo efficace.

````python
import sympy as sp

valore_esatto = sp.sin(sp.pi/4)*sp.cos(sp.pi/6) + sp.cos(sp.pi/4)*sp.sin(sp.pi/6)
print("Valore esatto:", sp.simplify(valore_esatto))
print("Confronto con sp.sin(75°):", sp.sin(sp.rad(75)))
print("Valore numerico:", float(valore_esatto))
````

## Riepilogo finale

- Formule di addizione: $\sin(\alpha\pm\beta)=\sin\alpha\cos\beta\pm\cos\alpha\sin\beta$; $\cos(\alpha\pm\beta)=\cos\alpha\cos\beta\mp\sin\alpha\sin\beta$.
- Formule di duplicazione (ponendo $\beta=\alpha$): $\sin(2\alpha)=2\sin\alpha\cos\alpha$; $\cos(2\alpha)=\cos^2\alpha-\sin^2\alpha=2\cos^2\alpha-1=1-2\sin^2\alpha$.
- Il coseno è una funzione pari ($\cos(-\theta)=\cos\theta$), il seno è dispari ($\sin(-\theta)=-\sin\theta$): da qui derivano le formule di sottrazione.
- Scomporre un angolo non notevole nella somma/differenza di due angoli notevoli permette di calcolarne seno e coseno esatti con le formule di addizione.
