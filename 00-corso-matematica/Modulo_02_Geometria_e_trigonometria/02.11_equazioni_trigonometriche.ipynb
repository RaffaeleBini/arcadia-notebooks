# 2.11 — Equazioni trigonometriche

**Prerequisiti:** 2.9 Seno, coseno, tangente, 2.10 Identità trigonometriche, 1.13 Equazioni esponenziali e logaritmiche.


## Perché ci interessa

Finora abbiamo calcolato seno e coseno di angoli **dati**. Ora ci poniamo la domanda inversa: per quali angoli $\theta$ vale, per esempio, $\sin\theta=\frac12$? Questo tipo di domanda — un'**equazione trigonometrica** — ha una particolarità importante rispetto alle equazioni algebriche viste finora: a causa della periodicità (notebook 2.8), ha quasi sempre **infinite soluzioni**.

## Teoria, passo dopo passo

### Equazioni elementari: $\sin\theta=k$

Per $|k|\le1$ (altrimenti nessuna soluzione, perché $\sin\theta\in[-1,1]$ sempre — è la coordinata $y$ di un punto sulla circonferenza goniometrica di raggio $1$, notebook 2.9), l'equazione $\sin\theta=k$ ha **due famiglie** di soluzioni nell'intervallo $[0,2\pi)$: chiamato $\theta_0$ un angolo con $\sin\theta_0=k$, l'altro è $\pi-\theta_0$ (per simmetria della circonferenza goniometrica rispetto all'asse $y$: due punti con la stessa ordinata $y$ sono simmetrici rispetto all'asse verticale).

A causa della periodicità (notebook 2.8: sommare $2k\pi$ non cambia il punto sulla circonferenza), la soluzione **generale** è:

$$\theta = \theta_0+2k\pi \quad\text{oppure}\quad \theta=(\pi-\theta_0)+2k\pi \qquad k\in\mathbb{Z}$$

### Equazioni elementari: $\cos\theta=k$

Analogamente, per $|k|\le1$: se $\theta_0$ soddisfa $\cos\theta_0=k$, l'altra famiglia è $-\theta_0$ (simmetria rispetto all'asse $x$: due punti con la stessa ascissa $x$ sono simmetrici rispetto all'asse orizzontale):

$$\theta=\pm\theta_0+2k\pi \qquad k\in\mathbb{Z}$$

### Equazioni elementari: $\tan\theta=k$

Per la tangente non c'è restrizione su $k$ (può essere qualunque numero reale, notebook 2.9). Se $\theta_0$ soddisfa $\tan\theta_0=k$, la periodicità della tangente è $\pi$ (non $2\pi$! perché $\tan(\theta+\pi)=\tan\theta$, essendo $\sin$ e $\cos$ entrambi cambiati di segno, e il loro rapporto resta invariato):

$$\theta=\theta_0+k\pi \qquad k\in\mathbb{Z}$$

### Equazioni riconducibili a quelle elementari

Molte equazioni si riconducono a una forma elementare tramite sostituzione o identità (notebook 2.10). Esempio: $\sin(2\theta)=\frac{\sqrt2}{2}$. Poniamo $\varphi=2\theta$ (sostituzione, come nel notebook 1.13): $\sin\varphi=\frac{\sqrt2}{2}$, che è elementare, con $\varphi_0=\frac{\pi}{4}$. Quindi $\varphi=\frac{\pi}{4}+2k\pi$ oppure $\varphi=\frac{3\pi}{4}+2k\pi$. Tornando a $\theta=\frac{\varphi}{2}$:

$$\theta = \frac{\pi}{8}+k\pi \quad\text{oppure}\quad \theta=\frac{3\pi}{8}+k\pi$$

(**attenzione**: dividendo per $2$ anche il periodo $2\pi$ diventa $\pi$: bisogna dividere *tutto*, non solo l'angolo principale).

### Equazioni con un solo tipo di funzione, di grado superiore

Se un'equazione contiene solo $\sin\theta$ (o solo $\cos\theta$), elevati a potenze diverse, si può trattare come un'equazione algebrica con la sostituzione $t=\sin\theta$ (o $t=\cos\theta$) — stessa idea del notebook 1.13 per le equazioni esponenziali. Esempio: $2\sin^2\theta-\sin\theta-1=0$ diventa $2t^2-t-1=0$ (notebook 1.6), risolvibile con la formula quadratica, per poi tornare a $\theta$ risolvendo $\sin\theta=t$ per ciascuna soluzione $t$ trovata (scartando quelle con $|t|>1$, impossibili per il seno).

````python
import sympy as sp

theta = sp.symbols('theta', real=True)

# Risolviamo sin(theta) = 1/2 con SymPy (restituisce la soluzione generale)
soluzione_generale = sp.solveset(sp.Eq(sp.sin(theta), sp.Rational(1, 2)), theta, domain=sp.S.Reals)
print("Soluzione generale di sin(theta)=1/2:")
sp.pprint(soluzione_generale)

# Soluzioni nell'intervallo [0, 2*pi)
import numpy as np
soluzioni_intervallo = [sp.pi/6, 5*sp.pi/6]
for s in soluzioni_intervallo:
    print(f"theta={s}: sin(theta)={sp.sin(s)}")

# Equazione con sostituzione: 2sin^2(theta) - sin(theta) - 1 = 0
t = sp.symbols('t')
soluzioni_t = sp.solve(sp.Eq(2*t**2 - t - 1, 0), t)
print("\nSoluzioni per t=sin(theta):", soluzioni_t)
for t_val in soluzioni_t:
    if abs(t_val) <= 1:
        sol = sp.solveset(sp.Eq(sp.sin(theta), t_val), theta, domain=sp.Interval(0, 2*sp.pi, False, True))
        print(f"  sin(theta)={t_val} -> theta in [0,2pi): {sol}")
    else:
        print(f"  sin(theta)={t_val}: scartata, |t|>1")
````

## Esercizio svolto e commentato

**Problema:** risolvi $\cos\theta=-\dfrac{\sqrt2}{2}$ nell'intervallo $[0,2\pi)$.

**Ragionamento:**

**Passo 1 — troviamo l'angolo di riferimento $\theta_0$** tale che $\cos\theta_0=\frac{\sqrt2}{2}$ (il valore assoluto): dalla tabella dei valori notevoli (notebook 2.9), $\theta_0=\frac{\pi}{4}$.

**Passo 2 — applichiamo lo schema per $\cos\theta=k$** (teoria): con $k=-\frac{\sqrt2}{2}$ negativo, cerchiamo dove il coseno vale esattamente $-\frac{\sqrt2}{2}$ (non $+\frac{\sqrt2}{2}$). Il coseno è negativo nel II e III quadrante (notebook 2.9). Usiamo la simmetria: gli angoli con $\cos\theta=-\frac{\sqrt2}{2}$ sono $\pi-\theta_0$ (II quadrante) e $\pi+\theta_0$ (III quadrante) — entrambi "riflessi" di $\theta_0$ rispetto a $\pi$.

$$\theta = \pi-\frac{\pi}{4}=\frac{3\pi}{4} \qquad\qquad \theta=\pi+\frac{\pi}{4}=\frac{5\pi}{4}$$

**Verifica**: $\cos\left(\frac{3\pi}{4}\right)=\cos\left(\pi-\frac{\pi}{4}\right)$. Usando la formula di sottrazione (notebook 2.10): $\cos\pi\cos\frac{\pi}{4}+\sin\pi\sin\frac{\pi}{4} = (-1)\cdot\frac{\sqrt2}{2}+0 = -\frac{\sqrt2}{2}$. ✓

Le soluzioni in $[0,2\pi)$ sono $\theta=\frac{3\pi}{4}$ e $\theta=\frac{5\pi}{4}$.

**Perché conviene sempre "portare" l'equazione a un angolo di riferimento positivo e poi ragionare sui quadranti, invece di cercare direttamente $\theta$ per il valore negativo:** la tabella dei valori notevoli (notebook 2.9) elenca solo valori positivi, per angoli del primo quadrante. Il metodo sistematico — trovare $\theta_0$ per il valore assoluto, poi usare la simmetria per posizionarsi nel quadrante giusto — funziona sempre, anche quando il valore dato non è "immediatamente" riconoscibile in una tabella.

````python
import sympy as sp

theta = sp.symbols('theta', real=True)
soluzioni = sp.solveset(sp.Eq(sp.cos(theta), -sp.sqrt(2)/2), theta, domain=sp.Interval(0, 2*sp.pi, False, True))
print("Soluzioni in [0,2π):", soluzioni)
````

## Riepilogo finale

- $\sin\theta=k$ (con $|k|\le1$) ha soluzione generale $\theta=\theta_0+2k\pi$ oppure $\theta=(\pi-\theta_0)+2k\pi$; $\cos\theta=k$ ha $\theta=\pm\theta_0+2k\pi$; $\tan\theta=k$ ha $\theta=\theta_0+k\pi$ (periodo $\pi$, non $2\pi$).
- Le equazioni trigonometriche hanno quasi sempre **infinite soluzioni**, a causa della periodicità (notebook 2.8): vanno sempre specificate rispetto a un intervallo, oppure espresse con il parametro intero $k$.
- Sostituzioni come $\varphi=n\theta$ riconducono equazioni più complesse a forme elementari, ma richiedono di dividere anche il periodo per $n$.
- Equazioni con una sola funzione trigonometrica elevata a potenze diverse si risolvono con sostituzione algebrica ($t=\sin\theta$ o $t=\cos\theta$), come nel notebook 1.13 per le esponenziali.
