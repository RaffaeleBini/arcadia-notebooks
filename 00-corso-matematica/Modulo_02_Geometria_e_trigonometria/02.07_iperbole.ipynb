# 2.7 — Iperbole

**Prerequisiti:** 2.6 Ellisse, 2.2 Retta.


## Perché ci interessa

Nel notebook 2.6 abbiamo definito l'ellisse imponendo che la **somma** delle distanze da due fuochi fosse costante. Una domanda naturale: cosa otteniamo imponendo invece che sia costante la loro **differenza** (in valore assoluto)? Il risultato è l'**iperbole**, l'ultima delle coniche "classiche" che studiamo in questo modulo, con una proprietà nuova rispetto alle curve precedenti: gli **asintoti**.

## Teoria, passo dopo passo

### Definizione come luogo geometrico

Fissati due fuochi $F_1,F_2$, l'iperbole è l'insieme dei punti $P$ tali che:

$$|PF_1-PF_2| = 2a \qquad (\text{costante, con } 0<2a<F_1F_2)$$

Il valore assoluto (notebook 0.8) è essenziale: non ci interessa quale distanza sia maggiore, solo che la loro differenza (in modulo) sia costante. Questo produce **due rami** separati della curva, uno per ciascun caso ($PF_1>PF_2$ oppure $PF_2>PF_1$).

### L'equazione canonica

Con gli stessi fuochi $F_1=(-c,0)$, $F_2=(c,0)$ del notebook 2.6, un procedimento analogo (formula della distanza, notebook 2.3) porta a:

$$\frac{x^2}{a^2}-\frac{y^2}{b^2}=1$$

Nota il **segno meno**, l'unica differenza formale rispetto all'ellisse. Qui però la relazione tra i parametri cambia: $b^2=c^2-a^2$ (invece di $a^2-c^2$) — conseguenza diretta del fatto che nell'iperbole $2a<F_1F_2=2c$, cioè $a<c$, mentre nell'ellisse $a>c$.

### Vertici e asintoti

Ponendo $y=0$: $x=\pm a$, i **vertici** dell'iperbole (uniche intersezioni con l'asse $x$; a differenza dell'ellisse, qui non c'è nessuna intersezione con l'asse $y$, perché ponendo $x=0$ si ottiene $-\frac{y^2}{b^2}=1$, impossibile per $y$ reale).

Il tratto distintivo dell'iperbole: per $|x|$ molto grande, la curva si avvicina sempre di più (senza mai toccarle) a due rette, le **asintote**:

$$y = \pm\frac{b}{a}x$$

Da dove vengono? Risolvendo l'equazione per $y$: $y=\pm b\sqrt{\frac{x^2}{a^2}-1} = \pm\frac{b}{a}\sqrt{x^2-a^2}$. Per $|x|$ molto grande rispetto ad $a$, il termine $-a^2$ sotto radice diventa via via meno rilevante rispetto a $x^2$ (un'idea intuitiva di limite, che formalizzeremo nel Modulo 4): $\sqrt{x^2-a^2}\approx\sqrt{x^2}=|x|$, quindi $y\approx\pm\frac{b}{a}x$: esattamente le rette asintote.

### Iperbole equilatera

Se $a=b$, l'equazione diventa $x^2-y^2=a^2$, e gli asintoti sono $y=\pm x$ (le bisettrici dei quadranti, perpendicolari tra loro, notebook 2.2 — coefficienti angolari $1$ e $-1$, il cui prodotto è $-1$). Questo caso speciale si chiama **iperbole equilatera**, e ha una forma particolarmente semplice se si scelgono come assi proprio gli asintoti: $xy=k$ (una funzione che incontreremo spesso, per esempio come modello di proporzionalità inversa, notebook 0.9).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

a_val, b_val = 3, 2
c_val = sp.sqrt(a_val**2 + b_val**2)
print(f"a={a_val}, b={b_val}, c={c_val} (relazione b^2=c^2-a^2 verificata:", b_val**2 == c_val**2 - a_val**2, ")")

# Grafico dell'iperbole e dei suoi asintoti
x_vals = np.linspace(-8, 8, 400)
x_vals = x_vals[np.abs(x_vals) >= a_val]  # dominio: |x|>=a

y_superiore = b_val * np.sqrt((x_vals/a_val)**2 - 1)
y_inferiore = -y_superiore

fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(x_vals, y_superiore, color='tab:blue')
ax.plot(x_vals, y_inferiore, color='tab:blue', label='iperbole')

x_asintoti = np.linspace(-8, 8, 10)
ax.plot(x_asintoti, (b_val/a_val)*x_asintoti, '--', color='gray', label='asintoti y=±(b/a)x')
ax.plot(x_asintoti, -(b_val/a_val)*x_asintoti, '--', color='gray')

ax.plot([-a_val, a_val], [0, 0], 'go', label='vertici')
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.set_aspect('equal')
ax.set_ylim(-8, 8)
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title(f"Iperbole x²/{a_val}²-y²/{b_val}²=1 e i suoi asintoti")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** un'iperbole ha equazione $\dfrac{x^2}{9}-\dfrac{y^2}{16}=1$. Trova vertici, fuochi e asintoti.

**Ragionamento:**

**Passo 1 — identifichiamo $a^2$ e $b^2$.** $a^2=9\Rightarrow a=3$; $b^2=16\Rightarrow b=4$.

**Passo 2 — troviamo i vertici**: $(\pm a,0)=(\pm3,0)$.

**Passo 3 — troviamo $c$**, usando $b^2=c^2-a^2$, cioè $c^2=a^2+b^2$ (attenzione: qui è un **più**, a differenza dell'ellisse):

$$c^2=9+16=25 \quad\Rightarrow\quad c=5$$

Fuochi: $F_1=(-5,0)$, $F_2=(5,0)$.

**Passo 4 — troviamo gli asintoti**:

$$y=\pm\frac{b}{a}x = \pm\frac{4}{3}x$$

**Verifica della definizione** sul vertice $(3,0)$: $PF_1=|3-(-5)|=8$, $PF_2=|3-5|=2$. Differenza: $|8-2|=6=2a$ (con $a=3$). ✓

**Perché la relazione tra $a$, $b$, $c$ ha il segno diverso rispetto all'ellisse, e come ricordarla senza confondersi:** nell'ellisse, $a$ è il semiasse **maggiore** (il più grande dei tre), quindi $b^2=a^2-c^2$ (una differenza tra il più grande e un altro termine, per restare positiva). Nell'iperbole, invece, $c$ è sempre il più grande (i fuochi sono "più lontani" dei vertici), quindi la relazione naturale è $c^2=a^2+b^2$ — la stessa struttura del teorema di Pitagora (notebook 2.3)! Non è un caso: $a$, $b$, $c$ nell'iperbole formano davvero un triangolo rettangolo, cosa che vale la pena ricordare per non confondere le due formule.

````python
import sympy as sp

a2, b2 = 9, 16
a, b = sp.sqrt(a2), sp.sqrt(b2)
c = sp.sqrt(a2 + b2)

print(f"a={a}, b={b}, c={c}")
print(f"Vertici: (±{a}, 0)")
print(f"Fuochi: (±{c}, 0)")
print(f"Asintoti: y = ±{b/a} x")

# Verifica definizione sul vertice
F1, F2 = (-c, 0), (c, 0)
P = (a, 0)
differenza = abs(abs(P[0]-F1[0]) - abs(P[0]-F2[0]))
print(f"\nVerifica: |PF1-PF2| = {differenza}, atteso 2a = {2*a}")
````

## Riepilogo finale

- Un'iperbole è il luogo dei punti la cui differenza (in modulo) delle distanze da due fuochi è costante ($=2a$).
- Equazione canonica: $\frac{x^2}{a^2}-\frac{y^2}{b^2}=1$, con $c^2=a^2+b^2$ (a differenza dell'ellisse, qui vale una relazione "alla Pitagora").
- L'iperbole ha due **asintoti**, $y=\pm\frac{b}{a}x$: rette a cui la curva si avvicina sempre di più senza mai toccarle.
- Se $a=b$ si parla di **iperbole equilatera**, con asintoti perpendicolari $y=\pm x$; scritta rispetto ai propri asintoti, diventa $xy=k$.
