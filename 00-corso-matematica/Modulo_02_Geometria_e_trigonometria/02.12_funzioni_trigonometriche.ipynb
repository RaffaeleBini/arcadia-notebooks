# 2.12 — Funzioni trigonometriche

**Prerequisiti:** 2.9 Seno, coseno, tangente, 2.11 Equazioni trigonometriche, 0.19 Relazioni e funzioni elementari.


## Perché ci interessa

Finora abbiamo pensato a $\sin\theta$ e $\cos\theta$ come a "coordinate di un punto sulla circonferenza goniometrica" (notebook 2.9), calcolate angolo per angolo. Ma sono anche, a tutti gli effetti, **funzioni** nel senso del notebook 0.19: $f(\theta)=\sin\theta$ associa a ogni $\theta$ reale un unico valore. In questo notebook studiamo il loro **grafico**, completando il quadro con le proprietà tipiche di una funzione: dominio, codominio, periodicità, simmetrie.

## Teoria, passo dopo passo

### Il grafico del seno

Se disegniamo $y=\sin\theta$ al variare di $\theta$ su tutto $\mathbb{R}$ (non solo $[0,2\pi)$), otteniamo un'onda che sale e scende regolarmente. Le proprietà principali:

- **Dominio**: tutto $\mathbb{R}$ (per ogni angolo esiste un punto sulla circonferenza goniometrica, notebook 2.8).
- **Codominio**: $[-1,1]$ (l'ordinata di un punto sulla circonferenza di raggio $1$ non può mai superare $1$ in valore assoluto).
- **Periodicità**: $\sin(\theta+2\pi)=\sin\theta$ per ogni $\theta$ (notebook 2.8): il grafico si ripete identico ogni $2\pi$.
- **Simmetria**: $\sin(-\theta)=-\sin\theta$ (notebook 2.10): funzione **dispari**, il grafico è simmetrico rispetto all'origine.
- **Zeri**: $\sin\theta=0$ per $\theta=k\pi$ ($k$ intero) — coerente con l'equazione elementare del notebook 2.11.

### Il grafico del coseno

Il grafico di $y=\cos\theta$ ha la stessa forma di quello del seno, ma **traslato**: infatti, dalla formula $\cos\theta=\sin\left(\theta+\frac{\pi}{2}\right)$ (verificabile con le formule di addizione, notebook 2.10), il grafico del coseno è quello del seno spostato di $\frac{\pi}{2}$ verso sinistra. Stesso dominio, codominio e periodo del seno, ma:

- **Simmetria**: $\cos(-\theta)=\cos\theta$: funzione **pari**, grafico simmetrico rispetto all'asse $y$.
- **Zeri**: $\cos\theta=0$ per $\theta=\frac{\pi}{2}+k\pi$.

### Il grafico della tangente

$y=\tan\theta$ ha un comportamento diverso, più simile a un'iperbole (notebook 2.7) ripetuta:

- **Dominio**: $\mathbb{R}\setminus\left\{\frac{\pi}{2}+k\pi\right\}$ (dove $\cos\theta=0$, la tangente non è definita, notebook 2.9).
- **Codominio**: tutto $\mathbb{R}$ (a differenza di seno e coseno, la tangente non è limitata: si avvicina a $+\infty$ e $-\infty$ vicino ai punti esclusi dal dominio, che sono **asintoti verticali**, un'idea analoga agli asintoti dell'iperbole, notebook 2.7).
- **Periodicità**: $\pi$ (non $2\pi$! come già osservato nel notebook 2.11).
- **Simmetria**: dispari, come il seno ($\tan(-\theta)=-\tan\theta$, rapporto di una funzione dispari e una pari).

### Perché la periodicità è la proprietà più importante da ricordare

Una funzione periodica di periodo $T$ soddisfa $f(\theta+T)=f(\theta)$ per ogni $\theta$: conoscere il comportamento della funzione su **un solo periodo** (per esempio $[0,2\pi)$ per seno e coseno) basta a conoscerlo ovunque. Questo è il motivo per cui, nel notebook 2.11, le equazioni trigonometriche si risolvono prima su un intervallo di un periodo, per poi generalizzare aggiungendo multipli del periodo.

````python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(-2*np.pi, 2*np.pi, 1000)

fig, axes = plt.subplots(3, 1, figsize=(9, 10))

axes[0].plot(theta, np.sin(theta), color='tab:blue')
axes[0].axhline(y=0, color='gray', linewidth=0.5)
axes[0].axhline(y=1, color='lightgray', linestyle='--')
axes[0].axhline(y=-1, color='lightgray', linestyle='--')
axes[0].set_title("y = sin(θ): dispari, periodo 2π, codominio [-1,1]")
axes[0].grid(True, alpha=0.3)

axes[1].plot(theta, np.cos(theta), color='tab:orange')
axes[1].axhline(y=0, color='gray', linewidth=0.5)
axes[1].set_title("y = cos(θ): pari, periodo 2π, codominio [-1,1]")
axes[1].grid(True, alpha=0.3)

# Per la tangente, evitiamo di disegnare vicino agli asintoti (altrimenti matplotlib traccia linee verticali spurie)
theta_tan = np.linspace(-2*np.pi, 2*np.pi, 2000)
tan_vals = np.tan(theta_tan)
tan_vals[np.abs(np.cos(theta_tan)) < 0.05] = np.nan  # spezza il grafico vicino agli asintoti
axes[2].plot(theta_tan, tan_vals, color='tab:green')
axes[2].axhline(y=0, color='gray', linewidth=0.5)
axes[2].set_ylim(-5, 5)
axes[2].set_title("y = tan(θ): dispari, periodo π, codominio R, asintoti verticali")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Verifica della relazione cos(theta) = sin(theta + pi/2)
import sympy as sp
t = sp.symbols('t')
print("cos(t) - sin(t+pi/2) =", sp.simplify(sp.cos(t) - sp.sin(t + sp.pi/2)))
````

## Esercizio svolto e commentato

**Problema:** studia la funzione $f(\theta)=2\sin(3\theta)$: trova codominio e periodo.

**Ragionamento:**

**Passo 1 — analizziamo l'effetto del fattore $2$ davanti al seno.** Il seno "puro" ha codominio $[-1,1]$ (teoria). Moltiplicare per $2$ scala tutti i valori: se $\sin(3\theta)\in[-1,1]$, allora $2\sin(3\theta)\in[-2,2]$. Il codominio è $[-2,2]$ (in generale, $A\sin(\dots)$ ha codominio $[-|A|,|A|]$; il numero $A$ si chiama **ampiezza**).

**Passo 2 — analizziamo l'effetto del fattore $3$ dentro il seno.** Il seno "puro" $\sin\varphi$ ha periodo $2\pi$: si ripete quando $\varphi$ aumenta di $2\pi$. Qui $\varphi=3\theta$: vogliamo trovare il periodo $T$ (in $\theta$) tale che $3(\theta+T)=3\theta+2\pi$, cioè $3T=2\pi$, quindi:

$$T=\frac{2\pi}{3}$$

Verifichiamo: $f(\theta+T)=2\sin\left(3\left(\theta+\frac{2\pi}{3}\right)\right)=2\sin(3\theta+2\pi)=2\sin(3\theta)=f(\theta)$. ✓ (usando la periodicità di base del seno, notebook 2.8).

**Conclusione**: $f(\theta)=2\sin(3\theta)$ ha codominio $[-2,2]$ e periodo $\frac{2\pi}{3}$: oscilla più velocemente (tre volte più velocemente) e più ampiamente (il doppio) del seno base.

**Perché il periodo si "restringe" dividendo per il coefficiente dentro la funzione (e non moltiplicando):** un coefficiente $3$ dentro l'argomento fa "correre" $\theta$ tre volte più in fretta verso il valore che completa un ciclo di $\sin$: serve un $\theta$ tre volte più piccolo per percorrere lo stesso ciclo. Questo tipo di ragionamento (in generale, $\sin(b\theta)$ ha periodo $\frac{2\pi}{|b|}$) è esattamente lo stesso schema usato nel notebook 2.11 per le equazioni con $\sin(2\theta)$.

````python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 1000)
f = 2 * np.sin(3*theta)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(theta, f)
ax.axhline(y=2, color='lightgray', linestyle='--')
ax.axhline(y=-2, color='lightgray', linestyle='--')
periodo = 2*np.pi/3
for k in range(4):
    ax.axvline(x=k*periodo, color='red', linestyle=':', linewidth=0.8)
ax.set_title(f"f(θ)=2sin(3θ): ampiezza 2, periodo {periodo:.4f} (linee rosse)")
ax.grid(True, alpha=0.3)
plt.show()
````

## Riepilogo finale

- $\sin\theta$ e $\cos\theta$ hanno dominio $\mathbb{R}$, codominio $[-1,1]$, periodo $2\pi$; il seno è dispari, il coseno è pari.
- $\tan\theta$ ha dominio $\mathbb{R}$ privato di $\frac{\pi}{2}+k\pi$, codominio $\mathbb{R}$, periodo $\pi$, ed è dispari, con asintoti verticali.
- In $f(\theta)=A\sin(b\theta)$, il numero $A$ (ampiezza) scala il codominio a $[-|A|,|A|]$, e il coefficiente $b$ comprime/allarga il periodo a $\frac{2\pi}{|b|}$.
- Conoscere una funzione periodica su un solo periodo basta a conoscerla ovunque: è il principio dietro la risoluzione delle equazioni trigonometriche (notebook 2.11).
