# 2.6 — Ellisse

**Prerequisiti:** 2.4 Circonferenza, 2.3 Distanza tra punti, 1.3 Scomposizione dei polinomi.


## Perché ci interessa

La circonferenza (notebook 2.4) è definita rispetto a **un** punto fisso, il centro. Cosa succede se, invece, definiamo un luogo geometrico rispetto a **due** punti fissi, chiedendo che la **somma** delle distanze da questi due punti sia costante? Otteniamo una nuova curva, l'**ellisse** — la forma delle orbite dei pianeti (come scoprì Keplero), e una generalizzazione naturale della circonferenza.

## Teoria, passo dopo passo

### Definizione come luogo geometrico

Fissiamo due punti $F_1$ e $F_2$, detti **fuochi**. L'ellisse è l'insieme dei punti $P$ tali che la somma delle distanze da $F_1$ e $F_2$ sia costante:

$$PF_1+PF_2 = 2a \qquad (\text{costante, con } 2a > F_1F_2)$$

(la condizione $2a>F_1F_2$ è necessaria: la somma di due lati di un "triangolo" $PF_1F_2$ deve superare il terzo lato, altrimenti il punto $P$ non esisterebbe in modo non degenere).

### L'equazione canonica

Se posizioniamo i fuochi sull'asse $x$, simmetrici rispetto all'origine, $F_1=(-c,0)$ e $F_2=(c,0)$, un calcolo con la formula della distanza (notebook 2.3, analogo nello spirito a quello della parabola) porta, dopo alcuni passaggi algebrici che qui non svolgiamo per esteso, all'equazione:

$$\frac{x^2}{a^2}+\frac{y^2}{b^2}=1$$

dove $b^2=a^2-c^2$ (una relazione che deriva direttamente dalla definizione, collegando i tre parametri).

### Il significato di $a$ e $b$

Ponendo $y=0$ nell'equazione: $\frac{x^2}{a^2}=1 \Rightarrow x=\pm a$: l'ellisse interseca l'asse $x$ nei punti $(\pm a,0)$, detti **vertici principali**. Ponendo $x=0$: $y=\pm b$, i **vertici secondari**. Il segmento tra i vertici principali si chiama **asse maggiore** (lunghezza $2a$), quello tra i vertici secondari **asse minore** (lunghezza $2b$, con $b<a$ se i fuochi sono sull'asse $x$).

**Perché il nome "canonica" richiede $a>b$ qui:** questa forma specifica presuppone i fuochi sull'asse $x$. Se invece $b>a$, i fuochi si trovano sull'asse $y$, e l'equazione (stessa forma algebrica!) descrive un'ellisse "più alta che larga" — la stessa equazione, letta con ruoli invertiti tra $a$ e $b$.

### La circonferenza come caso particolare

Cosa succede se $a=b$? L'equazione diventa $\frac{x^2}{a^2}+\frac{y^2}{a^2}=1$, cioè $x^2+y^2=a^2$: esattamente l'equazione di una **circonferenza** di raggio $a$ (notebook 2.4)! La circonferenza è quindi un caso particolare (degenere) dell'ellisse, quando i due fuochi coincidono ($c=0$, quindi $b^2=a^2-0=a^2$). Questa è la stessa relazione di generalizzazione che troveremo spesso in matematica: un concetto più semplice che riemerge come caso speciale di uno più generale.

### L'eccentricità

Il rapporto $e=\frac{c}{a}$ (con $0\le e<1$) si chiama **eccentricità**, e misura quanto l'ellisse è "schiacciata": $e=0$ corrisponde a una circonferenza perfetta ($c=0$); $e$ vicino a $1$ corrisponde a un'ellisse molto allungata.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Verifica della relazione b^2 = a^2 - c^2 con un esempio numerico
a_val, c_val = 5, 3
b_val = sp.sqrt(a_val**2 - c_val**2)
print(f"a={a_val}, c={c_val}, b={b_val}")

# Grafico dell'ellisse con questi parametri
theta = np.linspace(0, 2*np.pi, 200)
x_ellisse = a_val * np.cos(theta)
y_ellisse = float(b_val) * np.sin(theta)

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(x_ellisse, y_ellisse, label=f'x²/{a_val}²+y²/{float(b_val):.1f}²=1')
ax.plot([-c_val, c_val], [0, 0], 'ro', label='fuochi')
ax.plot([-a_val, a_val], [0, 0], 'go', markersize=5)
ax.plot([0, 0], [-float(b_val), float(b_val)], 'bo', markersize=5)
ax.axhline(y=0, color='gray', linewidth=0.5)
ax.axvline(x=0, color='gray', linewidth=0.5)
ax.set_aspect('equal')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title(f"Ellisse: eccentricità e = c/a = {c_val/a_val:.2f}")
plt.show()

# Verifica della definizione: somma delle distanze da un punto dell'ellisse ai due fuochi
def distanza(P, Q):
    return np.sqrt((P[0]-Q[0])**2 + (P[1]-Q[1])**2)

F1, F2 = (-c_val, 0), (c_val, 0)
for angolo in [0, np.pi/4, np.pi/2, np.pi]:
    P = (a_val*np.cos(angolo), float(b_val)*np.sin(angolo))
    somma = distanza(P, F1) + distanza(P, F2)
    print(f"Punto {P}: somma distanze dai fuochi = {somma:.4f}  (atteso 2a={2*a_val})")
````

## Esercizio svolto e commentato

**Problema:** un'ellisse ha equazione $\dfrac{x^2}{25}+\dfrac{y^2}{9}=1$. Trova i vertici, i fuochi, e l'eccentricità.

**Ragionamento:**

**Passo 1 — identifichiamo $a^2$ e $b^2$.** Confrontando con la forma canonica $\frac{x^2}{a^2}+\frac{y^2}{b^2}=1$: $a^2=25$, quindi $a=5$; $b^2=9$, quindi $b=3$. Poiché $a>b$, i fuochi sono sull'asse $x$ (coerente con la teoria).

**Passo 2 — troviamo i vertici.** Vertici principali: $(\pm a,0)=(\pm5,0)$. Vertici secondari: $(0,\pm b)=(0,\pm3)$.

**Passo 3 — troviamo $c$**, usando la relazione $b^2=a^2-c^2$, cioè $c^2=a^2-b^2$:

$$c^2=25-9=16 \quad\Rightarrow\quad c=4$$

I fuochi sono $F_1=(-4,0)$ e $F_2=(4,0)$.

**Passo 4 — calcoliamo l'eccentricità:**

$$e=\frac{c}{a}=\frac{4}{5}=0.8$$

Un valore relativamente vicino a $1$: l'ellisse è abbastanza "schiacciata" (non troppo simile a una circonferenza).

**Verifica della definizione** su un vertice principale, per esempio $(5,0)$: $PF_1=|5-(-4)|=9$, $PF_2=|5-4|=1$. Somma: $9+1=10=2a$ (con $a=5$). ✓

**Perché verifichiamo proprio su un vertice, e non su un punto qualsiasi:** i vertici sono i punti più semplici da calcolare a mano (le distanze diventano differenze dirette sull'asse $x$, senza bisogno della formula generale del notebook 2.3). È una buona pratica: quando possibile, verifica un risultato nel caso più semplice disponibile, dove il calcolo è meno soggetto a errori.

````python
import sympy as sp

a2, b2 = 25, 9
a, b = sp.sqrt(a2), sp.sqrt(b2)
c = sp.sqrt(a2 - b2)
eccentricita = c / a

print(f"a={a}, b={b}, c={c}")
print(f"Vertici principali: (±{a}, 0)")
print(f"Vertici secondari: (0, ±{b})")
print(f"Fuochi: (±{c}, 0)")
print(f"Eccentricità: {eccentricita} = {float(eccentricita)}")

# Verifica sul vertice (5,0)
F1, F2 = (-c, 0), (c, 0)
P = (a, 0)
somma = abs(P[0]-F1[0]) + abs(P[0]-F2[0])
print(f"\nVerifica: PF1+PF2 = {somma}, atteso 2a = {2*a}")
````

## Riepilogo finale

- Un'ellisse è il luogo dei punti la cui somma delle distanze da due fuochi $F_1,F_2$ è costante ($=2a$).
- Equazione canonica (fuochi sull'asse $x$): $\frac{x^2}{a^2}+\frac{y^2}{b^2}=1$, con $b^2=a^2-c^2$ e $c$ la semidistanza tra i fuochi.
- La circonferenza (notebook 2.4) è il caso particolare $a=b$ (fuochi coincidenti, $c=0$).
- L'**eccentricità** $e=\frac{c}{a}\in[0,1)$ misura quanto l'ellisse è schiacciata: $e=0$ è una circonferenza perfetta, $e$ vicino a $1$ un'ellisse molto allungata.
