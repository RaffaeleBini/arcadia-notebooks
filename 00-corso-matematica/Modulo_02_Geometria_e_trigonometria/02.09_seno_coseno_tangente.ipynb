# 2.9 — Seno, coseno, tangente

**Prerequisiti:** 2.8 Angoli, 2.3 Distanza tra punti.


## Perché ci interessa

Nel notebook 2.8 abbiamo imparato a misurare un angolo sulla circonferenza goniometrica, individuando un punto sul cerchio. Ma quali sono le **coordinate** di quel punto? Questa domanda, apparentemente semplice, è la porta d'ingresso alla trigonometria: le risposte si chiamano **coseno** e **seno**, e da queste due funzioni deriva (quasi) tutto il resto.

## Teoria, passo dopo passo

### Definizione sulla circonferenza goniometrica

Dato un angolo $\theta$ (misurato come nel notebook 2.8), consideriamo il punto $P$ sulla circonferenza goniometrica individuato da quell'angolo. Per definizione:

$$\cos\theta = x_P \qquad \sin\theta = y_P$$

cioè: il **coseno** è l'ascissa di $P$, il **seno** è la sua ordinata. Poiché $P$ sta sulla circonferenza $x^2+y^2=1$ (notebook 2.8), otteniamo immediatamente la relazione più importante di tutta la trigonometria:

$$\cos^2\theta+\sin^2\theta=1$$

(questa è solo l'equazione della circonferenza goniometrica, riscritta con $x=\cos\theta$, $y=\sin\theta$: non è un teorema misterioso, è la definizione stessa).

### Collegamento con il triangolo rettangolo

Per angoli tra $0$ e $\frac{\pi}{2}$ (primo quadrante), c'è un'interpretazione equivalente e forse più familiare: in un triangolo rettangolo con un angolo $\theta$, $\cos\theta$ è il rapporto tra il cateto adiacente e l'ipotenusa, $\sin\theta$ tra il cateto opposto e l'ipotenusa. Questo coincide con la definizione sulla circonferenza goniometrica: se disegniamo il triangolo rettangolo con ipotenusa il raggio (lunghezza $1$) e vertice in $P$, i cateti hanno proprio lunghezza $|x_P|=|\cos\theta|$ e $|y_P|=|\sin\theta|$ (formula della distanza, notebook 2.3, applicata al segmento da $P$ agli assi).

### La tangente

La **tangente** è definita come il rapporto:

$$\tan\theta = \frac{\sin\theta}{\cos\theta} \qquad (\cos\theta\ne0)$$

Geometricamente, $\tan\theta$ è il coefficiente angolare (notebook 2.2) della retta che unisce l'origine al punto $P$! Infatti, il coefficiente angolare di una retta per l'origine e per $(x_P,y_P)$ è $\frac{y_P}{x_P}=\frac{\sin\theta}{\cos\theta}$. La tangente non è definita quando $\cos\theta=0$ (frazione algebrica, notebook 1.4: denominatore nullo), cioè per $\theta=\frac{\pi}{2}+k\pi$.

### Valori notevoli

Alcuni angoli hanno valori di seno e coseno esprimibili con radicali semplici, che vale la pena memorizzare (li useremo continuamente):

| $\theta$ | $0$ | $\frac{\pi}{6}$ | $\frac{\pi}{4}$ | $\frac{\pi}{3}$ | $\frac{\pi}{2}$ |
|---|---|---|---|---|---|
| $\sin\theta$ | $0$ | $\frac12$ | $\frac{\sqrt2}{2}$ | $\frac{\sqrt3}{2}$ | $1$ |
| $\cos\theta$ | $1$ | $\frac{\sqrt3}{2}$ | $\frac{\sqrt2}{2}$ | $\frac12$ | $0$ |

Osserviamo una simmetria: i valori di $\sin$ e $\cos$ si "scambiano" passando da $\frac{\pi}{6}$ a $\frac{\pi}{3}$ (e $\frac{\pi}{4}$ è il punto di scambio, dove sono uguali). Non è un caso: $\frac{\pi}{6}$ e $\frac{\pi}{3}$ sono **angoli complementari** (sommano a $\frac{\pi}{2}$), e vale sempre $\sin\theta=\cos\left(\frac{\pi}{2}-\theta\right)$ (conseguenza diretta della simmetria della circonferenza goniometrica rispetto alla bisettrice $y=x$).

### Segni nei quattro quadranti

Il segno di $\cos\theta$ e $\sin\theta$ dipende dal quadrante (notebook 2.1) in cui cade il punto $P$: nel I quadrante entrambi positivi; nel II, $\cos<0<\sin$; nel III, entrambi negativi; nel IV, $\sin<0<\cos$. Il segno di $\tan\theta$ segue dalla regola dei segni della divisione (notebook 0.2): positivo nel I e III quadrante, negativo nel II e IV.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Verifica dell'identità fondamentale su vari angoli
angoli = [0, sp.pi/6, sp.pi/4, sp.pi/3, sp.pi/2, sp.pi, 3*sp.pi/2]
for a in angoli:
    identita = sp.simplify(sp.cos(a)**2 + sp.sin(a)**2)
    print(f"theta={a}: cos²+sin² = {identita}")

# Tabella dei valori notevoli
print("\nValori notevoli:")
for a in [0, sp.pi/6, sp.pi/4, sp.pi/3, sp.pi/2]:
    print(f"  sin({a}) = {sp.sin(a)},  cos({a}) = {sp.cos(a)}")

# Disegno sulla circonferenza goniometrica: seno e coseno come coordinate
theta_val = np.pi/3
fig, ax = plt.subplots(figsize=(6, 6))
theta_cerchio = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(theta_cerchio), np.sin(theta_cerchio), color='lightgray')
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)

x_P, y_P = np.cos(theta_val), np.sin(theta_val)
ax.plot([0, x_P], [0, y_P], 'b-')
ax.plot(x_P, y_P, 'ro')
ax.plot([x_P, x_P], [0, y_P], 'g--', label=f'sin(60°)={y_P:.3f}')
ax.plot([0, x_P], [0, 0], 'orange', linestyle='--', label=f'cos(60°)={x_P:.3f}')
ax.set_aspect('equal')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("theta=60°: cos è l'ascissa, sin è l'ordinata di P")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** sapendo che $\theta$ è nel II quadrante e $\sin\theta=\dfrac35$, trova $\cos\theta$ e $\tan\theta$.

**Ragionamento:**

**Passo 1 — usiamo l'identità fondamentale** $\cos^2\theta+\sin^2\theta=1$ per trovare $\cos^2\theta$:

$$\cos^2\theta = 1-\sin^2\theta = 1-\left(\frac35\right)^2 = 1-\frac{9}{25}=\frac{16}{25}$$

**Passo 2 — estraiamo la radice quadrata** (notebook 0.6), ricordando che ci sono **due** possibilità matematicamente, $\cos\theta=\pm\frac45$:

$$\cos\theta = \pm\sqrt{\frac{16}{25}} = \pm\frac45$$

**Passo 3 — usiamo l'informazione sul quadrante per scegliere il segno corretto.** Nel II quadrante (teoria, sopra): $\cos\theta<0$. Quindi scartiamo la soluzione positiva:

$$\cos\theta = -\frac45$$

**Passo 4 — calcoliamo la tangente:**

$$\tan\theta = \frac{\sin\theta}{\cos\theta} = \frac{3/5}{-4/5} = -\frac34$$

(coerente con la regola dei segni: II quadrante, $\tan\theta$ negativa).

**Perché l'identità fondamentale da sola non basta, e serve l'informazione sul quadrante:** l'equazione $\cos^2\theta+\sin^2\theta=1$ è un'equazione di secondo grado in $\cos\theta$ (notebook 1.6): come ogni equazione di questo tipo, può avere due soluzioni. Senza sapere in quale quadrante si trova $\theta$, non potremmo distinguere quale delle due sia quella giusta — l'informazione sul quadrante è indispensabile, non un dettaglio superfluo.

````python
import sympy as sp

sin_theta = sp.Rational(3, 5)
cos_theta_quadro = 1 - sin_theta**2
cos_theta_positivo = sp.sqrt(cos_theta_quadro)
print(f"cos²θ = {cos_theta_quadro}")
print(f"cos θ = ±{cos_theta_positivo}")

# Nel II quadrante cos è negativo
cos_theta = -cos_theta_positivo
tan_theta = sin_theta / cos_theta
print(f"\nScegliendo il segno per il II quadrante: cos θ = {cos_theta}")
print(f"tan θ = {tan_theta}")
````

## Riepilogo finale

- $\cos\theta$ e $\sin\theta$ sono, per definizione, le coordinate del punto sulla circonferenza goniometrica individuato dall'angolo $\theta$: $\cos\theta=x_P$, $\sin\theta=y_P$.
- L'identità fondamentale $\cos^2\theta+\sin^2\theta=1$ è semplicemente l'equazione della circonferenza goniometrica.
- $\tan\theta=\frac{\sin\theta}{\cos\theta}$ è il coefficiente angolare della retta OP; non definita quando $\cos\theta=0$.
- Il segno di seno, coseno e tangente dipende dal quadrante; l'identità fondamentale da sola non distingue il segno, serve sempre l'informazione sul quadrante.
