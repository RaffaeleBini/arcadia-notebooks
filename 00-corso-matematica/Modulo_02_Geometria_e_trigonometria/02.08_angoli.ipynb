## Perché ci interessa

Finora abbiamo misurato gli angoli, se mai lo abbiamo fatto, in **gradi**: un angolo giro è $360°$. Ma i gradi sono una scelta arbitraria (perché $360$ e non $100$, o $1000$?). Per fare matematica in modo rigoroso — e in particolare per collegare angoli e numeri reali, come serve alla trigonometria — ci serve un'unità di misura **naturale**, legata direttamente alla geometria del cerchio: il **radiante**.

Questo notebook introduce gli angoli sulla circonferenza (notebook 2.4), preparando il terreno per definire seno, coseno e tangente nel prossimo notebook.

## Teoria, passo dopo passo

### La circonferenza goniometrica

D'ora in avanti lavoreremo quasi sempre con la **circonferenza goniometrica**: la circonferenza di centro l'origine e raggio $1$ (caso particolare del notebook 2.4, con $x_0=y_0=0$, $r=1$):

$$x^2+y^2=1$$

Un angolo si misura a partire dal **semiasse positivo delle $x$**, ruotando in senso **antiorario** (la direzione convenzionalmente positiva) fino a raggiungere un raggio della circonferenza. Un angolo negativo corrisponde a una rotazione in senso orario.

### La misura in radianti

Il **radiante** misura un angolo tramite la lunghezza dell'**arco** di circonferenza goniometrica che l'angolo "sottende" (racchiude). Poiché la circonferenza goniometrica ha raggio $1$, la sua lunghezza totale è $2\pi\times1=2\pi$ (dalla formula della lunghezza della circonferenza, $2\pi r$). Un angolo giro completo misura quindi **$2\pi$ radianti**.

Questa scelta non è arbitraria come i gradi: è **intrinseca** alla geometria del cerchio, non dipende da una convenzione culturale (come il $360$ babilonese). Per questo il radiante è l'unità "naturale" degli angoli, e sarà l'unica che useremo a partire dal Modulo 5 (derivate), dove semplifica enormemente le formule.

### Conversione tra gradi e radianti

Poiché $360°$ corrispondono a $2\pi$ radianti, la proporzione (notebook 0.9) è:

$$\frac{\text{gradi}}{360} = \frac{\text{radianti}}{2\pi} \qquad\Longrightarrow\qquad \text{radianti} = \text{gradi}\times\frac{\pi}{180}$$

Alcuni valori che vale la pena memorizzare (li useremo costantemente):

| Gradi | $0°$ | $30°$ | $45°$ | $60°$ | $90°$ | $180°$ | $270°$ | $360°$ |
|---|---|---|---|---|---|---|---|---|
| Radianti | $0$ | $\frac{\pi}{6}$ | $\frac{\pi}{4}$ | $\frac{\pi}{3}$ | $\frac{\pi}{2}$ | $\pi$ | $\frac{3\pi}{2}$ | $2\pi$ |

### Angoli e periodicità

Ruotare di $2\pi$ radianti (un giro completo) riporta esattamente allo stesso raggio di partenza. Questo significa che angoli che differiscono per un multiplo intero di $2\pi$ individuano **lo stesso punto** sulla circonferenza goniometrica: $\theta$ e $\theta+2k\pi$ (per $k$ intero, notebook 0.2) sono, in questo senso, "lo stesso angolo". Questa osservazione — la **periodicità** — sarà centrale quando definiremo le funzioni trigonometriche (notebook 2.9 e 2.12): saranno funzioni che si ripetono sempre uguali ogni $2\pi$.

````python
import numpy as np
import matplotlib.pyplot as plt

# Conversione gradi <-> radianti
def gradi_a_radianti(g):
    return g * np.pi / 180

def radianti_a_gradi(r):
    return r * 180 / np.pi

angoli_gradi = [0, 30, 45, 60, 90, 180, 270, 360]
for g in angoli_gradi:
    r = gradi_a_radianti(g)
    print(f"{g}° = {r:.4f} rad", end="")
    if g in [30, 45, 60, 90, 180, 270, 360]:
        # confronto con la forma esatta con pi greco
        frazioni = {30: "π/6", 45: "π/4", 60: "π/3", 90: "π/2", 180: "π", 270: "3π/2", 360: "2π"}
        print(f"  (= {frazioni[g]})")
    else:
        print()

# Disegniamo la circonferenza goniometrica con alcuni angoli notevoli
fig, ax = plt.subplots(figsize=(6, 6))
theta_cerchio = np.linspace(0, 2*np.pi, 200)
ax.plot(np.cos(theta_cerchio), np.sin(theta_cerchio), color='tab:blue')
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)

for g in [0, 30, 45, 60, 90, 180, 270]:
    r = gradi_a_radianti(g)
    x, y = np.cos(r), np.sin(r)
    ax.plot([0, x], [0, y], color='gray', linewidth=0.8)
    ax.plot(x, y, 'ro')
    ax.annotate(f"{g}°", (x*1.15, y*1.15))

ax.set_aspect('equal')
ax.set_title("Circonferenza goniometrica con alcuni angoli notevoli")
ax.grid(True, alpha=0.3)
plt.show()
````

## Esercizio svolto e commentato

**Problema:** converti $135°$ in radianti (in forma esatta, con $\pi$), e trova un angolo negativo equivalente (stesso raggio sulla circonferenza) minore di $-180°$.

**Ragionamento:**

**Passo 1 — conversione in radianti**, usando la formula $\text{radianti}=\text{gradi}\times\frac{\pi}{180}$:

$$135\times\frac{\pi}{180} = \frac{135\pi}{180}$$

Semplifichiamo la frazione $\frac{135}{180}$ trovando il massimo comun divisore (notebook 0.3): $135=27\times5$, $180=36\times5$, quindi $\frac{135}{180}=\frac{27}{36}=\frac{3}{4}$ (dividendo ancora per $9$). Il risultato è:

$$135° = \frac{3\pi}{4} \text{ radianti}$$

**Passo 2 — troviamo un angolo equivalente negativo minore di $-180°$.** Per la periodicità (teoria), sottraendo $360°$ (un giro completo) otteniamo un angolo che individua lo stesso raggio:

$$135°-360° = -225°$$

Verifichiamo che sia minore di $-180°$: sì, $-225°<-180°$. ✓ (Se avessimo bisogno di scendere ancora, potremmo sottrarre altri $360°$, ottenendo $-585°$, e così via — infiniti angoli equivalenti, tutti della forma $135°+k\times360°$.)

**Perché "sottrarre $360°$" produce sempre un angolo che individua lo stesso punto sulla circonferenza:** per definizione, un giro completo ($360°=2\pi$ radianti) riporta esattamente al punto di partenza. Sottrarre (o aggiungere) un multiplo intero di $360°$ equivale a fare uno o più giri completi in più o in meno, senza cambiare la posizione finale sul cerchio — è esattamente l'osservazione sulla periodicità vista nella teoria.

````python
import sympy as sp

gradi = 135
radianti = sp.Rational(gradi, 180) * sp.pi
print(f"{gradi}° in radianti:", radianti)

angolo_equivalente = gradi - 360
print(f"Angolo equivalente: {angolo_equivalente}°  (minore di -180°? {angolo_equivalente < -180})")
````
