## Perché ci interessa

Nel notebook 2.1 abbiamo visto che $y=x$ descrive una retta a $45°$ passante per l'origine. Ma quella era solo un caso particolare. In questo notebook cerchiamo l'equazione **generale** di una retta qualsiasi nel piano, e scopriamo che è, letteralmente, un'equazione di primo grado in due incognite (notebook 1.5) — lo stesso oggetto algebrico, letto ora come un disegno.

## Teoria, passo dopo passo

### Dalla proporzionalità alla forma esplicita

Immaginiamo una retta che sale in modo "costante": ogni volta che $x$ aumenta di $1$, $y$ aumenta sempre della stessa quantità $m$. Questo rapporto costante si chiama **coefficiente angolare** (o pendenza):

$$m = \frac{\Delta y}{\Delta x} = \frac{y_2-y_1}{x_2-x_1}$$

per due punti qualsiasi $(x_1,y_1)$, $(x_2,y_2)$ sulla retta (con $x_1\ne x_2$). Il fatto che $m$ sia lo stesso per **qualunque** coppia di punti della retta è precisamente la definizione geometrica di "retta": un tasso di variazione costante — la stessa idea di proporzionalità diretta vista nel notebook 0.9, applicata qui al rapporto tra variazioni.

L'equazione di una retta non verticale si scrive nella **forma esplicita**:

$$y = mx+q$$

dove $m$ è il coefficiente angolare e $q$ è l'**intercetta**, il valore di $y$ dove la retta incrocia l'asse $y$ (cioè quando $x=0$: infatti $y=m\cdot0+q=q$).

### Il segno di $m$ e il caso $m=0$

- $m>0$: la retta **sale** da sinistra a destra.
- $m<0$: la retta **scende** da sinistra a destra.
- $m=0$: $y=q$ per ogni $x$, una retta **orizzontale**.

Le rette **verticali** (come $x=3$, tutti i punti con ascissa $3$) non si possono scrivere nella forma $y=mx+q$ (il coefficiente angolare sarebbe "infinito": $\Delta x=0$ renderebbe $\frac{\Delta y}{\Delta x}$ non definito, notebook 0.3). Si scrivono semplicemente come $x=k$.

### La forma implicita

Più in generale, ogni retta (comprese le verticali) si può scrivere nella **forma implicita**:

$$ax+by+c=0$$

con $a,b$ non entrambi nulli. Se $b\ne0$, possiamo risolvere per $y$ (invariantiva, notebook 0.5) e tornare alla forma esplicita: $y=-\frac{a}{b}x-\frac{c}{b}$, quindi $m=-\frac{a}{b}$. Se $b=0$ (e $a\ne0$), l'equazione diventa $x=-\frac{c}{a}$: una retta verticale.

### Retta per due punti

Dati due punti $(x_1,y_1)$ e $(x_2,y_2)$, con $x_1\ne x_2$, calcoliamo prima $m$ con la formula vista sopra, poi usiamo uno dei due punti per trovare $q$: sostituendo $(x_1,y_1)$ nell'equazione $y=mx+q$, otteniamo $q=y_1-mx_1$.

### Rette parallele e perpendicolari

Due rette $y=m_1x+q_1$ e $y=m_2x+q_2$ sono:

- **Parallele** se e solo se $m_1=m_2$ (stesso "tasso di salita/discesa": non si incontrano mai, a meno che non siano la stessa retta — coerente con il notebook 1.7, dove un sistema con coefficienti proporzionali e termini noti diversi era impossibile).
- **Perpendicolari** se e solo se $m_1\cdot m_2=-1$ (equivalentemente, $m_2=-\frac{1}{m_1}$: il coefficiente angolare "reciproco e opposto"). Non dimostriamo qui questa condizione (richiede un po' di trigonometria, che vedremo nei prossimi notebook), ma la useremo spesso.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y, m, q = sp.symbols('x y m q')

# Coefficiente angolare tra due punti
def coefficiente_angolare(p1, p2):
    return sp.Rational(p2[1] - p1[1], p2[0] - p1[0])

m_calcolato = coefficiente_angolare((1, 2), (3, 6))
print("Coefficiente angolare tra (1,2) e (3,6):", m_calcolato)

# Retta per due punti: troviamo m e q
p1, p2 = (1, 2), (3, 6)
m_val = coefficiente_angolare(p1, p2)
q_val = p1[1] - m_val * p1[0]
print(f"Equazione della retta: y = {m_val}x + {q_val}")

# Grafico di rette parallele e perpendicolari
x_vals = np.linspace(-3, 5, 100)
fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(x_vals, 2*x_vals + 1, label='y=2x+1  (m=2)')
ax.plot(x_vals, 2*x_vals - 2, label='y=2x-2  (m=2, parallela)')
ax.plot(x_vals, -0.5*x_vals + 1, label='y=-0.5x+1  (m=-1/2, perpendicolare alla prima)')
ax.axhline(y=0, color='black', linewidth=0.8)
ax.axvline(x=0, color='black', linewidth=0.8)
ax.legend()
ax.set_aspect('equal')
ax.set_ylim(-6, 6)
ax.grid(True, alpha=0.3)
ax.set_title("Rette parallele (stesso m) e perpendicolari (m1*m2=-1)")
plt.show()

# Verifica condizione di perpendicolarità
print("\nm1*m2 =", 2 * (-0.5), "(deve essere -1)")
````

## Esercizio svolto e commentato

**Problema:** trova l'equazione della retta passante per $(2,-1)$ e perpendicolare alla retta $y=3x+5$.

**Ragionamento:**

**Passo 1 — troviamo il coefficiente angolare della retta perpendicolare.** La retta data ha $m_1=3$. Per la condizione di perpendicolarità (notebook 2.2, teoria): $m_1\cdot m_2=-1$, quindi:

$$m_2 = -\frac{1}{m_1} = -\frac{1}{3}$$

**Passo 2 — usiamo il punto dato per trovare $q$.** La retta cercata ha equazione $y=-\frac13x+q$, e deve passare per $(2,-1)$: sostituiamo $x=2$, $y=-1$ (proprio come sostituire un valore in un'equazione, notebook 1.5):

$$-1 = -\frac13(2)+q \quad\Rightarrow\quad -1=-\frac23+q \quad\Rightarrow\quad q=-1+\frac23=-\frac13$$

**Passo 3 — scriviamo l'equazione finale:**

$$y=-\frac13x-\frac13$$

**Verifica**: il punto $(2,-1)$ soddisfa l'equazione? $-\frac13(2)-\frac13=-\frac23-\frac13=-1$. ✓ E la condizione di perpendicolarità? $3\times\left(-\frac13\right)=-1$. ✓

**Perché "sostituire il punto dato nell'equazione generica" è la strategia giusta per trovare $q$:** conosciamo la forma dell'equazione (a meno di $q$) grazie alla condizione di perpendicolarità, ma ci manca un'informazione (la posizione esatta della retta). Il punto dato è esattamente l'informazione che manca: imponendo che l'equazione sia vera per quel punto, otteniamo un'equazione di primo grado nell'unica incognita rimasta, $q$ (notebook 1.5).

````python
import sympy as sp

x, y, qs = sp.symbols('x y q')
m1 = 3
m2 = sp.Rational(-1, m1)

equazione = sp.Eq(-1, m2*2 + qs)
q_trovato = sp.solve(equazione, qs)[0]
print(f"m2 = {m2}, q = {q_trovato}")
print(f"Equazione della retta: y = {m2}x + {q_trovato}")
````

**4.** Coefficiente angolare della retta per $(0,0)$ e $(4,2)$: $m=\dfrac{2-0}{4-0}=\frac12$. La retta cercata ha lo stesso $m$ (parallela) e passa per $(1,3)$: $3=\frac12(1)+q \Rightarrow q=3-\frac12=\frac52$. Equazione: $y=\frac12x+\frac52$.

## Riepilogo finale

- L'equazione esplicita di una retta è $y=mx+q$: $m$ è il coefficiente angolare (costante per ogni coppia di punti della retta), $q$ è l'intercetta con l'asse $y$.
- Ogni retta (comprese le verticali) si scrive in forma implicita $ax+by+c=0$; le rette verticali si scrivono $x=k$ e non hanno forma esplicita.
- Due rette sono **parallele** se hanno lo stesso $m$; sono **perpendicolari** se $m_1m_2=-1$.
- Per trovare l'equazione di una retta con condizioni parziali (un punto e una pendenza, o due punti), si sostituisce quanto noto nell'equazione generica e si risolve per l'incognita rimasta (notebook 1.5).
