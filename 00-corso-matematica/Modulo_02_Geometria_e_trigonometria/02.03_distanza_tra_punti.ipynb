# 2.3 — Distanza tra punti

**Prerequisiti:** 2.1 Piano cartesiano, 0.8 Valore assoluto, 0.6 Frazioni, potenze e radici.


## Perché ci interessa

Nel notebook 0.8 abbiamo definito la distanza tra due numeri sulla retta come $|a-b|$. Ma nel piano cartesiano, due punti generici non stanno quasi mai sulla stessa retta orizzontale o verticale: come misuriamo la distanza tra, per esempio, $(1,2)$ e $(4,6)$?

La risposta è uno strumento antichissimo applicato in modo nuovo: il **teorema di Pitagora**. Questa formula ci servirà in tutto il resto del modulo, per definire circonferenze, ellissi, e per calcolare lunghezze in generale.

## Teoria, passo dopo passo

### Costruire la formula con Pitagora

Consideriamo due punti $A=(x_1,y_1)$ e $B=(x_2,y_2)$. Costruiamo un terzo punto ausiliario $C=(x_2,y_1)$: ha la stessa ascissa di $B$ e la stessa ordinata di $A$. Per come lo abbiamo costruito (notebook 2.1: stessa ordinata → stessa orizzontale; stessa ascissa → stessa verticale), il segmento $AC$ è **orizzontale** e il segmento $CB$ è **verticale**: sono perpendicolari, e formano con $AB$ un triangolo rettangolo con l'ipotenusa $AB$.

Le lunghezze dei due cateti si calcolano come distanze sulla retta (notebook 0.8, valore assoluto):

$$AC = |x_2-x_1| \qquad CB=|y_2-y_1|$$

Per il **teorema di Pitagora** (il quadrato dell'ipotenusa è la somma dei quadrati dei cateti):

$$AB^2 = AC^2+CB^2 = |x_2-x_1|^2+|y_2-y_1|^2 = (x_2-x_1)^2+(y_2-y_1)^2$$

(abbiamo usato $|a|^2=a^2$, notebook 0.8: il quadrato di un valore assoluto è uguale al quadrato del numero, perché elevare al quadrato "cancella" comunque il segno). Estraendo la radice quadrata (notebook 0.6, sempre il valore non negativo, coerente con una lunghezza):

$$AB = \sqrt{(x_2-x_1)^2+(y_2-y_1)^2}$$

Questa è la **formula della distanza** tra due punti del piano.

### Verifica sui casi già noti

Se $A$ e $B$ hanno la stessa ordinata ($y_1=y_2$), la formula si riduce a $AB=\sqrt{(x_2-x_1)^2}=|x_2-x_1|$ (notebook 0.6, $\sqrt{a^2}=|a|$): esattamente la distanza sulla retta orizzontale, coerente con quanto sapevamo già. Analogamente per punti con la stessa ascissa.

### Il punto medio

Un'altra quantità utile, collegata alla distanza: il **punto medio** $M$ del segmento $AB$ ha coordinate che sono le **medie** delle coordinate di $A$ e $B$:

$$M = \left(\frac{x_1+x_2}{2},\ \frac{y_1+y_2}{2}\right)$$

Questo segue dall'idea di "stare a metà strada" sia orizzontalmente sia verticalmente: la media di due numeri (notebook 0.3, densità di $\mathbb{Q}$) è sempre il punto intermedio tra loro.

````python
import sympy as sp
import matplotlib.pyplot as plt

def distanza(A, B):
    return sp.sqrt((B[0]-A[0])**2 + (B[1]-A[1])**2)

A, B = (1, 2), (4, 6)
d = distanza(A, B)
print(f"Distanza tra {A} e {B}: {d}")

# Verifica con il punto ausiliario C, per mostrare visivamente il triangolo rettangolo
C = (B[0], A[1])
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(*zip(A, B, C, A), 'o-')
for nome, punto in zip(["A", "B", "C"], [A, B, C]):
    ax.annotate(f"{nome}{punto}", punto, textcoords="offset points", xytext=(8, 5))
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title(f"Triangolo rettangolo ausiliario: AB = {float(d):.3f}")
plt.show()

# Punto medio
M = ((A[0]+B[0])/2, (A[1]+B[1])/2)
print(f"Punto medio tra {A} e {B}: {M}")
````

## Esercizio svolto e commentato

**Problema:** i punti $A=(-1,1)$, $B=(3,1)$, $C=(3,4)$ sono vertici di un triangolo. Calcola il perimetro e stabilisci se è rettangolo.

**Ragionamento:**

**Passo 1 — calcoliamo le tre distanze.** Notiamo che $A$ e $B$ condividono l'ordinata ($y=1$): $AB=|3-(-1)|=4$ (distanza diretta, senza radice). $B$ e $C$ condividono l'ascissa ($x=3$): $BC=|4-1|=3$. Per $AC$ serve la formula generale:

$$AC = \sqrt{(3-(-1))^2+(4-1)^2} = \sqrt{4^2+3^2} = \sqrt{16+9}=\sqrt{25}=5$$

**Passo 2 — calcoliamo il perimetro:**

$$P = AB+BC+CA = 4+3+5=12$$

**Passo 3 — verifichiamo se è rettangolo**, controllando il teorema di Pitagora al contrario: il lato più lungo ($AC=5$) dovrebbe essere l'ipotenusa se il triangolo è rettangolo, cioè dovrebbe valere $AC^2=AB^2+BC^2$:

$$5^2 = 4^2+3^2 \quad\Rightarrow\quad 25=16+9=25 \ \checkmark$$

Sì, è un triangolo rettangolo, con l'angolo retto in $B$ (dove si incontrano i due cateti $AB$ e $BC$).

**Perché potevamo aspettarcelo ancora prima di calcolare:** $A$ e $B$ hanno la stessa ordinata (lato $AB$ orizzontale), $B$ e $C$ hanno la stessa ascissa (lato $BC$ verticale) — esattamente lo schema del notebook 2.1, dove un lato orizzontale e uno verticale che condividono un vertice garantiscono un angolo retto lì, senza bisogno di calcoli. Il calcolo con Pitagora qui serve a trovare le **lunghezze** esatte, non a stabilire che l'angolo è retto (quello lo sapevamo già dalla struttura dei punti).

````python
import sympy as sp

A, B, C = (-1, 1), (3, 1), (3, 4)

def distanza(P, Q):
    return sp.sqrt((Q[0]-P[0])**2 + (Q[1]-P[1])**2)

AB = distanza(A, B)
BC = distanza(B, C)
CA = distanza(C, A)
print(f"AB={AB}, BC={BC}, CA={CA}")
print(f"Perimetro: {AB+BC+CA}")
print(f"Pitagora: AC^2 = AB^2+BC^2 ?", CA**2 == AB**2 + BC**2)
````

## Riepilogo finale

- La distanza tra $A=(x_1,y_1)$ e $B=(x_2,y_2)$ è $AB=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}$, ottenuta applicando il teorema di Pitagora a un triangolo rettangolo ausiliario.
- Se i due punti condividono un'ascissa o un'ordinata, la formula si riduce alla semplice distanza $|a-b|$ vista nel notebook 0.8.
- Il punto medio tra $A$ e $B$ ha come coordinate le medie delle coordinate di $A$ e $B$.
- Un lato orizzontale e uno verticale che condividono un vertice garantiscono un angolo retto in quel vertice, senza bisogno di calcoli — utile per riconoscere triangoli rettangoli a colpo d'occhio.
