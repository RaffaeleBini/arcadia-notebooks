# 2.4 — Circonferenza

**Prerequisiti:** 2.3 Distanza tra punti, 1.2 Prodotti notevoli, 1.6 Equazioni di secondo grado.


## Perché ci interessa

Una circonferenza è, per definizione, l'insieme di tutti i punti che stanno a una distanza fissa (il **raggio**) da un punto fisso (il **centro**). Nel notebook 2.3 abbiamo appena costruito la formula per la distanza tra due punti: possiamo usarla direttamente per tradurre questa definizione geometrica in un'**equazione algebrica** — il nostro primo vero esempio di "luogo geometrico" scritto in formula, anticipato nel notebook 2.1.

## Teoria, passo dopo passo

### Dall'equidistanza all'equazione

Sia $C=(x_0,y_0)$ il centro e $r>0$ il raggio. Un punto generico $P=(x,y)$ appartiene alla circonferenza se e solo se la sua distanza da $C$ è esattamente $r$ (definizione di circonferenza, nessuna sorpresa):

$$\sqrt{(x-x_0)^2+(y-y_0)^2} = r$$

Eleviamo al quadrato entrambi i membri (lecito: entrambi non negativi, come nell'esercizio del notebook 2.3):

$$(x-x_0)^2+(y-y_0)^2 = r^2$$

Questa è l'**equazione della circonferenza in forma canonica** (o "forma centro-raggio"). Basta leggerla per riconoscere subito centro e raggio.

### La forma generale

Sviluppando i quadrati con i prodotti notevoli (notebook 1.2):

$$x^2-2x_0x+x_0^2+y^2-2y_0y+y_0^2=r^2$$

Riordinando e chiamando $a=-2x_0$, $b=-2y_0$, $c=x_0^2+y_0^2-r^2$, otteniamo la **forma generale**:

$$x^2+y^2+ax+by+c=0$$

Ogni circonferenza si può scrivere così, ma non ogni equazione di questa forma è una circonferenza "vera" (lo vedremo tra poco). Per tornare dalla forma generale a quella canonica (trovare centro e raggio), si usa il **completamento del quadrato** — la stessa tecnica algebrica usata nel notebook 1.6 per ricavare la formula risolutiva delle equazioni di secondo grado:

$$x_0=-\frac{a}{2} \qquad y_0=-\frac{b}{2} \qquad r=\sqrt{x_0^2+y_0^2-c}$$

### Quando l'equazione generale NON è una vera circonferenza

Perché serva $r>0$ (un raggio reale e positivo), deve risultare $x_0^2+y_0^2-c\ge0$. Tre casi, in analogia diretta con il discriminante delle equazioni di secondo grado (notebook 1.6):

- Se $x_0^2+y_0^2-c>0$: **circonferenza vera**, di raggio $r=\sqrt{x_0^2+y_0^2-c}>0$.
- Se $x_0^2+y_0^2-c=0$: **circonferenza degenere**, un solo punto (il centro stesso, raggio $0$).
- Se $x_0^2+y_0^2-c<0$: **nessun punto reale** soddisfa l'equazione (non esiste raggio reale il cui quadrato sia negativo, notebook 0.4).

### Intersezione tra retta e circonferenza

Per trovare dove una retta (notebook 2.2) interseca una circonferenza, si mette a **sistema** (notebook 1.7) l'equazione della retta e quella della circonferenza: sostituendo l'espressione di $y$ (dalla retta) nell'equazione della circonferenza si ottiene un'equazione di secondo grado in $x$ sola (notebook 1.6), il cui discriminante $\Delta$ dice quante intersezioni ci sono: $\Delta>0$ due punti (retta **secante**), $\Delta=0$ un punto (retta **tangente**), $\Delta<0$ nessun punto (retta **esterna**).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y')

# Dalla forma generale alla forma canonica, con completamento del quadrato
a, b, c = -4, -6, 4  # esempio: x^2+y^2-4x-6y+4=0
x0, y0 = -a/2, -b/2
raggio_quadro = x0**2 + y0**2 - c
print(f"Centro: ({x0},{y0})")
print(f"r^2 = {raggio_quadro}")
if raggio_quadro > 0:
    print(f"Raggio: {sp.sqrt(raggio_quadro)}")

# Verifica con SymPy: risolviamo l'equazione generale per confermare centro/raggio
equazione_generale = x**2 + y**2 + a*x + b*y + c
forma_canonica = sp.expand((x - x0)**2 + (y - y0)**2 - raggio_quadro)
print("\nCoincide con la forma generale sviluppata?", sp.expand(equazione_generale - forma_canonica) == 0)

# Grafico della circonferenza
theta = np.linspace(0, 2*np.pi, 200)
raggio = float(sp.sqrt(raggio_quadro))
x_cerchio = x0 + raggio * np.cos(theta)
y_cerchio = y0 + raggio * np.sin(theta)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x_cerchio, y_cerchio)
ax.plot(x0, y0, 'ko')
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title(f"x²+y²-4x-6y+4=0  ->  centro ({x0},{y0}), raggio {raggio}")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova i punti di intersezione tra la retta $y=x+1$ e la circonferenza $x^2+y^2=25$.

**Ragionamento:**

**Passo 1 — impostiamo il sistema** (notebook 1.7):

$$\begin{cases} y=x+1 \\ x^2+y^2=25 \end{cases}$$

**Passo 2 — sostituiamo** l'espressione di $y$ dalla prima equazione nella seconda (metodo di sostituzione, notebook 1.7):

$$x^2+(x+1)^2=25$$

**Passo 3 — sviluppiamo** il quadrato del binomio (notebook 1.2) e semplifichiamo:

$$x^2+x^2+2x+1=25 \quad\Rightarrow\quad 2x^2+2x+1-25=0 \quad\Rightarrow\quad 2x^2+2x-24=0$$

Dividiamo tutto per $2$ (invariantiva, notebook 0.5): $x^2+x-12=0$.

**Passo 4 — risolviamo l'equazione di secondo grado** (notebook 1.6): $\Delta=1+48=49>0$, quindi due soluzioni:

$$x = \frac{-1\pm7}{2} \quad\Rightarrow\quad x_1=3,\ x_2=-4$$

**Passo 5 — troviamo le $y$ corrispondenti**, usando $y=x+1$: per $x_1=3$, $y_1=4$; per $x_2=-4$, $y_2=-3$.

I punti di intersezione sono $(3,4)$ e $(-4,-3)$.

**Verifica**: entrambi devono soddisfare $x^2+y^2=25$: $3^2+4^2=9+16=25$ ✓; $(-4)^2+(-3)^2=16+9=25$ ✓.

**Perché il discriminante $\Delta$ della equazione risultante ci dice il tipo di intersezione, ancora prima di calcolare le soluzioni:** avendo trovato $\Delta=49>0$, sapevamo già (notebook 1.6) che ci sarebbero state due soluzioni distinte per $x$, quindi due punti di intersezione: la retta è **secante** alla circonferenza. Se avessimo trovato $\Delta=0$, la retta sarebbe stata tangente; $\Delta<0$, esterna — senza bisogno di completare il calcolo per saperlo.

````python
import sympy as sp

x, y = sp.symbols('x y')
soluzioni = sp.solve([sp.Eq(y, x + 1), sp.Eq(x**2 + y**2, 25)], [x, y])
print("Punti di intersezione:", soluzioni)
````

## Riepilogo finale

- La circonferenza di centro $(x_0,y_0)$ e raggio $r$ ha equazione canonica $(x-x_0)^2+(y-y_0)^2=r^2$, diretta conseguenza della formula della distanza (notebook 2.3).
- Sviluppando i quadrati si ottiene la forma generale $x^2+y^2+ax+by+c=0$; il completamento del quadrato (come nel notebook 1.6) permette di tornare a centro e raggio.
- L'equazione generale rappresenta una vera circonferenza solo se $x_0^2+y_0^2-c>0$ (altrimenti un punto singolo, o nessun punto reale).
- L'intersezione tra retta e circonferenza si trova mettendo a sistema (notebook 1.7); il discriminante dell'equazione risultante (notebook 1.6) distingue retta secante, tangente o esterna.
