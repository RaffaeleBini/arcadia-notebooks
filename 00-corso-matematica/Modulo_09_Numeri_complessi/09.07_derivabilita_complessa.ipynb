# 9.7 — Derivabilità complessa

**Prerequisiti:** 9.6 Funzioni olomorfe, 7.7 Derivate parziali, 7.8 Differenziabilità.


## Perché ci interessa

Nel notebook 9.6 abbiamo verificato la $\mathbb C$-derivabilità caso per caso, controllando esplicitamente il limite lungo diverse direzioni — un metodo che funziona, ma è macchinoso e non dà un criterio pratico. Questo notebook fornisce esattamente quel criterio: le **equazioni di Cauchy-Riemann**, che collegano la $\mathbb C$-derivabilità alle derivate parziali (Modulo 7) della parte reale $u$ e immaginaria $v$. È il ponte che permette di verificare l'olomorfia con un calcolo diretto, senza limiti direzionali.

## Teoria, passo dopo passo

### Derivare lungo due direzioni specifiche

Sia $f(z)=u(x,y)+iv(x,y)$ derivabile in $z_0=x_0+iy_0$. Calcoliamo il limite del rapporto incrementale lungo **due** direzioni specifiche (che devono dare lo **stesso** risultato, essendo $f$ derivabile per ipotesi).

**Lungo l'asse reale** ($h=t\in\mathbb R$, $t\to0$):

$$f'(z_0) = \lim_{t\to0}\frac{f(x_0+t+iy_0)-f(x_0+iy_0)}t = \lim_{t\to0}\frac{u(x_0+t,y_0)-u(x_0,y_0)}t+i\lim_{t\to0}\frac{v(x_0+t,y_0)-v(x_0,y_0)}t$$

Riconosciamo esattamente le **derivate parziali** rispetto a $x$ (notebook 7.7): $f'(z_0)=\dfrac{\partial u}{\partial x}(x_0,y_0)+i\dfrac{\partial v}{\partial x}(x_0,y_0)$.

**Lungo l'asse immaginario** ($h=it$, $t\in\mathbb R$, $t\to0$):

$$f'(z_0) = \lim_{t\to0}\frac{f(x_0+i(y_0+t))-f(x_0+iy_0)}{it} = \frac1i\left(\frac{\partial u}{\partial y}(x_0,y_0)+i\frac{\partial v}{\partial y}(x_0,y_0)\right)$$

Usando $\dfrac1i=-i$ (verificabile: $\dfrac1i\times i=1$, e $-i\times i=-i^2=1$): $f'(z_0)=-i\dfrac{\partial u}{\partial y}(x_0,y_0)+\dfrac{\partial v}{\partial y}(x_0,y_0)=\dfrac{\partial v}{\partial y}(x_0,y_0)-i\dfrac{\partial u}{\partial y}(x_0,y_0)$.

### Le equazioni di Cauchy-Riemann

Uguagliando le due espressioni di $f'(z_0)$ (parte reale con parte reale, parte immaginaria con parte immaginaria, notebook 9.1):

$$\boxed{\frac{\partial u}{\partial x} = \frac{\partial v}{\partial y} \qquad\qquad \frac{\partial u}{\partial y} = -\frac{\partial v}{\partial x}}$$

Queste sono le **equazioni di Cauchy-Riemann** (C-R): una **condizione necessaria** per la $\mathbb C$-derivabilità in $z_0$. Se falliscono in un punto, $f$ **non** è $\mathbb C$-derivabile lì (esattamente come abbiamo verificato "manualmente" per $\bar z$ e $\operatorname{Re}(z)$ nel notebook 9.6).

### Condizione sufficiente

> **Teorema.** Se $u,v$ hanno derivate parziali continue in un intorno di $(x_0,y_0)$ (cioè sono differenziabili in senso reale, notebook 7.8) e soddisfano le equazioni di Cauchy-Riemann in quel punto, allora $f=u+iv$ è $\mathbb C$-derivabile in $z_0$, con $f'(z_0)=\dfrac{\partial u}{\partial x}+i\dfrac{\partial v}{\partial x}$.

Questo teorema (che non dimostriamo nel dettaglio, ma la cui idea centrale è che la differenziabilità reale, notebook 7.8, garantisce che il resto $o(\|h\|)$ sia "isotropo" — lo stesso in ogni direzione — e le C-R garantiscano che la parte lineare si comporti in modo consistente con la moltiplicazione complessa) rende le C-R un criterio **completo**: derivate parziali continue + C-R $\iff$ olomorfa.

### Riesame del controesempio $\bar z$

Per $f(z)=\bar z=x-iy$: $u=x,v=-y$. $\partial u/\partial x=1$, $\partial v/\partial y=-1$: **diverse**! La prima equazione di Cauchy-Riemann fallisce ovunque: coerente con quanto trovato "a mano" nel notebook 9.6.

### Cauchy-Riemann e il Laplaciano

Se $u,v$ hanno derivate parziali seconde continue, derivando la prima equazione C-R rispetto a $x$ e la seconda rispetto a $y$: $\partial^2u/\partial x^2=\partial^2v/\partial y\partial x$ e $\partial^2u/\partial y^2=-\partial^2v/\partial x\partial y$. Sommando, e usando il teorema di Schwarz (notebook 7.10, le derivate miste coincidono): $\dfrac{\partial^2u}{\partial x^2}+\dfrac{\partial^2u}{\partial y^2}=0$, cioè $\Delta u=0$ (notebook 7.21): **la parte reale di ogni funzione olomorfa è armonica**! (E lo stesso vale per $v$, con un calcolo analogo.) Questo collega l'analisi complessa alla teoria del potenziale, un'anticipazione del Modulo 15.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

def cauchy_riemann(u, v):
    eq1 = sp.simplify(sp.diff(u, x) - sp.diff(v, y))
    eq2 = sp.simplify(sp.diff(u, y) + sp.diff(v, x))
    return eq1, eq2

# f(z) = z^2: u = x^2-y^2, v = 2xy
u1, v1 = x**2 - y**2, 2*x*y
print("f(z)=z^2, Cauchy-Riemann:", cauchy_riemann(u1, v1), "-> (0,0) significa soddisfatte ovunque")

# f(z) = conjugate(z): u = x, v = -y
u2, v2 = x, -y
print("f(z)=conjugate(z), Cauchy-Riemann:", cauchy_riemann(u2, v2), "-> non nulle, C-R fallisce")

# Verifica che u=x^2-y^2 è armonica (Laplaciano nullo)
laplaciano_u1 = sp.diff(u1, x, 2) + sp.diff(u1, y, 2)
print("\nLaplaciano di u=x^2-y^2:", laplaciano_u1, "-> armonica, coerente con f(z)=z^2 olomorfa")
````

## Esercizio svolto e commentato

**Problema:** verifica se $f(z)=e^z$ soddisfa le equazioni di Cauchy-Riemann, e conferma la formula $f'(z)=e^z$ trovata nel notebook 9.6.

**Ragionamento:**

**Passo 1 — scriviamo $u,v$** (dal notebook 9.5): $u(x,y)=e^x\cos y$, $v(x,y)=e^x\sin y$.

**Passo 2 — calcoliamo $\partial u/\partial x$**: $\partial u/\partial x=e^x\cos y$ (derivando l'esponenziale, $\cos y$ costante rispetto a $x$).

**Passo 3 — calcoliamo $\partial v/\partial y$**: $\partial v/\partial y=e^x\cos y$ (derivando $\sin y$ rispetto a $y$, $e^x$ costante).

**Passo 4 — confrontiamo**: $\partial u/\partial x=e^x\cos y=\partial v/\partial y$. **Prima equazione C-R soddisfatta.**

**Passo 5 — calcoliamo $\partial u/\partial y$ e $-\partial v/\partial x$**: $\partial u/\partial y=-e^x\sin y$ (derivando $\cos y$). $\partial v/\partial x=e^x\sin y$, quindi $-\partial v/\partial x=-e^x\sin y$. **Coincidono: seconda equazione C-R soddisfatta.**

**Passo 6 — applichiamo la formula per $f'(z)$** (teoria): $f'(z)=\dfrac{\partial u}{\partial x}+i\dfrac{\partial v}{\partial x}=e^x\cos y+ie^x\sin y=e^x(\cos y+i\sin y)$.

**Conclusione:** riconosciamo, per la formula di Eulero (notebook 9.3) applicata a $e^x(\cos y+i\sin y)$, che questo è esattamente $e^{x+iy}=e^z$: $f'(z)=e^z$, confermando (ora tramite Cauchy-Riemann, non più solo tramite il limite diretto) il risultato del notebook 9.6.

**Perché verificare le C-R è, in pratica, sempre più rapido che calcolare il limite direzionale del rapporto incrementale (come fatto nel notebook 9.6):** le equazioni di Cauchy-Riemann riducono il problema della $\mathbb C$-derivabilità a un controllo di **quattro derivate parziali** (calcolabili con le tecniche standard del notebook 7.7) e un confronto algebrico — un procedimento meccanico e sempre applicabile — mentre verificare il limite lungo "tutte le direzioni possibili" richiederebbe, in linea di principio, infiniti controlli (o un argomento generale ad hoc, come quello usato per $\bar z$). Le C-R trasformano un problema di analisi (limiti) in un problema di calcolo differenziale elementare.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
u = sp.exp(x) * sp.cos(y)
v = sp.exp(x) * sp.sin(y)

eq1 = sp.simplify(sp.diff(u, x) - sp.diff(v, y))
eq2 = sp.simplify(sp.diff(u, y) + sp.diff(v, x))
print("du/dx - dv/dy =", eq1)
print("du/dy + dv/dx =", eq2)

f_prime = sp.diff(u, x) + sp.I * sp.diff(v, x)
print("\nf'(z) = du/dx + i*dv/dx =", sp.simplify(f_prime))
````

## Riepilogo finale

- Le **equazioni di Cauchy-Riemann** $\partial u/\partial x=\partial v/\partial y$ e $\partial u/\partial y=-\partial v/\partial x$ sono condizione **necessaria** per la $\mathbb C$-derivabilità, ottenute confrontando il rapporto incrementale lungo l'asse reale e quello immaginario.
- Con derivate parziali **continue**, le C-R diventano anche **sufficienti**: un criterio completo per l'olomorfia.
- Se $f$ è olomorfa (con derivate seconde continue), sia $u$ sia $v$ sono **armoniche** ($\Delta u=\Delta v=0$, notebook 7.21) — un collegamento profondo con la teoria del potenziale.
- Le C-R trasformano la verifica dell'olomorfia da un problema di limiti direzionali a un calcolo diretto di derivate parziali.
