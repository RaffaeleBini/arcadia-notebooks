# 9.6 — Funzioni olomorfe

**Prerequisiti:** 9.5 Funzioni complesse, 5.8 Derivata, 9.1 Forma cartesiana.


## Perché ci interessa

Nel notebook 5.8 abbiamo definito la derivata $f'(x_0)=\lim_{h\to0}\frac{f(x_0+h)-f(x_0)}h$ per funzioni reali, con $h$ un numero reale che tende a $0$ da sinistra o da destra. Per una funzione complessa $f:\mathbb C\to\mathbb C$, possiamo scrivere la **stessa identica formula**, ma ora $h$ è un numero complesso che può tendere a $0$ da **infinite** direzioni (esattamente come i limiti in più variabili del notebook 7.5). Questa apparente piccola generalizzazione ha conseguenze enormi: la derivabilità complessa risulta una condizione **molto più forte** della differenziabilità reale, e le funzioni che la soddisfano — le funzioni **olomorfe** — hanno proprietà quasi miracolose, che studieremo per il resto di questo modulo.

## Teoria, passo dopo passo

### La definizione (identica nella forma al caso reale)

Sia $f:D\to\mathbb C$, $D\subseteq\mathbb C$ aperto (notebook 7.3, la stessa nozione di insieme aperto si applica identicamente a $\mathbb C$, identificato con $\mathbb R^2$). $f$ è **derivabile in senso complesso** (o $\mathbb C$-derivabile) in $z_0\in D$ se esiste il limite:

$$f'(z_0) = \lim_{h\to0}\frac{f(z_0+h)-f(z_0)}h$$

dove ora $h\in\mathbb C$ tende a $0$ (nel senso della norma complessa, cioè del modulo, notebook 9.1) — **non** solo lungo l'asse reale o immaginario, ma lungo **qualsiasi** direzione o traiettoria nel piano complesso, esattamente come nel notebook 7.5.

### Funzione olomorfa

$f$ si dice **olomorfa** (o analitica) in un punto $z_0$ se è $\mathbb C$-derivabile non solo in $z_0$ ma in un'intera **palla aperta** attorno a $z_0$ (notebook 7.3) — cioè, derivabile "in un intorno", non solo puntualmente. Una funzione olomorfa su tutto $\mathbb C$ si dice **intera**.

### Le regole di derivazione si trasferiscono immediatamente

Poiché la definizione è algebricamente identica al caso reale, **tutte** le regole di derivazione del notebook 5.9 (somma, prodotto, quoziente, catena) si dimostrano con **la stessa identica dimostrazione**, semplicemente sostituendo $x\in\mathbb R$ con $z\in\mathbb C$: $(f+g)'=f'+g'$, $(fg)'=f'g+fg'$, $(z^n)'=nz^{n-1}$, ecc. Per esempio, ogni **polinomio** $p(z)=a_nz^n+\cdots+a_0$ è una funzione **intera**, con $p'(z)=na_nz^{n-1}+\cdots+a_1$, esattamente come nel caso reale.

### L'esponenziale complesso è olomorfo

Verifichiamo $\left(e^z\right)'=e^z$ direttamente dalla definizione: $\dfrac{e^{z+h}-e^z}h=e^z\dfrac{e^h-1}h$ (usando $e^{z+h}=e^ze^h$, notebook 9.5). Per $h\to0$ (in $\mathbb C$), si dimostra (con la serie di potenze del notebook 9.3) che $\dfrac{e^h-1}h\to1$ — lo stesso limite notevole del caso reale, ora esteso a $h$ complesso. Quindi $(e^z)'=e^z\times1=e^z$: l'esponenziale complesso è **intero**, con la stessa formula di derivazione del caso reale.

### Un controesempio cruciale: $f(z)=\bar z$ NON è derivabile

Consideriamo $f(z)=\bar z$ (il coniugio, notebook 9.1). Calcoliamo il rapporto incrementale in $z_0=0$: $\dfrac{f(0+h)-f(0)}h=\dfrac{\bar h-0}h=\dfrac{\bar h}h$. Proviamo $h$ **reale** (tendente a $0$ lungo l'asse reale): $\bar h=h$ (un numero reale è il proprio coniugato), quindi $\dfrac{\bar h}h=1$. Proviamo $h$ **immaginario puro**, $h=it$ con $t\to0$: $\bar h=\overline{it}=-it$, quindi $\dfrac{\bar h}h=\dfrac{-it}{it}=-1$. **Due direzioni diverse danno limiti diversi** ($1$ e $-1$): esattamente il fenomeno del notebook 7.5! Il limite non esiste: $f(z)=\bar z$ **non è $\mathbb C$-derivabile in nessun punto**, nonostante sia una funzione "innocua" — anzi, come vedremo nel prossimo notebook, addirittura differenziabile in senso reale (Modulo 7). Questo mostra concretamente perché la derivabilità complessa è una condizione così più severa: deve reggere lungo **infinite** direzioni contemporaneamente.

````python
import sympy as sp

h = sp.symbols('h')
z = sp.symbols('z')

# Verifica: (e^z)' = e^z, tramite il limite del rapporto incrementale
rapporto = (sp.exp(z + h) - sp.exp(z)) / h
limite = sp.limit(rapporto, h, 0)
print("Limite del rapporto incrementale per e^z:", limite)

# Controesempio: f(z) = conjugate(z), rapporto incrementale lungo direzioni diverse
t = sp.symbols('t', real=True)

# h reale, h -> 0
h_reale = t
rapporto_reale = sp.conjugate(h_reale) / h_reale
print("\nRapporto con h reale:", rapporto_reale)

# h immaginario puro, h -> 0
h_imm = sp.I * t
rapporto_imm = sp.conjugate(h_imm) / h_imm
print("Rapporto con h immaginario puro:", sp.simplify(rapporto_imm))

print("\nDue limiti diversi (1 e -1): il limite del rapporto incrementale NON esiste")
print("=> f(z)=conjugate(z) NON è C-derivabile in nessun punto")
````

## Esercizio svolto e commentato

**Problema:** verifica, usando le regole di derivazione trasferite dal caso reale, che $f(z)=z^3+2z$ è una funzione intera, e calcola $f'(z)$; poi calcola $f'(1+i)$.

**Ragionamento:**

**Passo 1 — riconosciamo che $f$ è un polinomio**, quindi (per la teoria) è olomorfa su tutto $\mathbb C$: una funzione **intera**.

**Passo 2 — applichiamo le regole di derivazione**, identiche al caso reale (notebook 5.9): $\dfrac d{dz}(z^3)=3z^2$ (regola della potenza) e $\dfrac d{dz}(2z)=2$ (derivata di una funzione lineare). Per linearità della derivata: $f'(z)=3z^2+2$.

**Passo 3 — valutiamo in $z_0=1+i$**: prima calcoliamo $(1+i)^2=1+2i+i^2=1+2i-1=2i$ (quadrato di binomio, notebook 1.2, con $i^2=-1$).

**Passo 4 — sostituiamo**: $f'(1+i)=3\times2i+2=6i+2=2+6i$.

**Conclusione:** $f'(z)=3z^2+2$, e $f'(1+i)=2+6i$.

**Perché possiamo applicare le regole di derivazione del notebook 5.9 "senza modifiche", senza dover riverificare ogni singola dimostrazione nel caso complesso:** le dimostrazioni delle regole di derivazione (somma, prodotto, potenza) nel notebook 5.9 usano solo proprietà algebriche generali — distributività, associatività, il fatto che il limite di una somma è la somma dei limiti — che valgono **identicamente** in $\mathbb C$ (essendo $\mathbb C$ un campo, notebook 9.1, con le stesse proprietà algebriche di $\mathbb R$, più la struttura di limite basata sul modulo, che si comporta come il valore assoluto). Non c'è nulla nella dimostrazione originale che dipenda specificamente dall'ordinamento di $\mathbb R$ (che $\mathbb C$ non possiede) — quindi le regole si trasferiscono di peso, senza bisogno di nuove dimostrazioni.

````python
import sympy as sp

z = sp.symbols('z')
f = z**3 + 2*z
f_prime = sp.diff(f, z)
print("f'(z) =", f_prime)

valore = f_prime.subs(z, 1 + sp.I)
print("f'(1+i) =", sp.expand(valore, complex=True))
````

## Riepilogo finale

- Una funzione $f:\mathbb C\to\mathbb C$ è **$\mathbb C$-derivabile** in $z_0$ se il limite del rapporto incrementale esiste, indipendentemente dalla direzione di avvicinamento di $h\to0$ nel piano complesso — una condizione molto più forte del caso reale.
- Una funzione derivabile in un intorno di ogni suo punto si dice **olomorfa**; su tutto $\mathbb C$, **intera**.
- Tutte le regole di derivazione del notebook 5.9 si trasferiscono senza modifiche; polinomi ed esponenziale complesso sono funzioni intere.
- $f(z)=\bar z$ (e altre funzioni "semplici" come $\operatorname{Re}(z)$) **non sono mai** $\mathbb C$-derivabili: un fenomeno senza analogo nel caso reale.
