# 9.14 — Teorema dei residui

**Prerequisiti:** 9.13 Residui, 9.9 Teorema di Cauchy, 5.18 Integrali impropri.


## Perché ci interessa

Nel notebook 9.13 abbiamo visto che $\oint_\gamma f\,dz=2\pi i\operatorname{Res}(f,z_0)$ per un cerchio attorno a **una sola** singolarità. Il **teorema dei residui** generalizza questo fatto a curve che circondano **più** singolarità contemporaneamente, sommando i contributi di ciascuna. È il risultato culminante dell'analisi complessa elementare, e — sorprendentemente — diventa uno strumento pratico per calcolare **integrali reali** (notebook 5.15, 5.18) che nessuna tecnica del Modulo 5 riusciva a risolvere in forma chiusa.

## Teoria, passo dopo passo

### Enunciato

> **Teorema dei residui.** Sia $f$ olomorfa su un dominio semplicemente connesso $D$, tranne che in un numero finito di singolarità isolate $z_1,\dots,z_k\in D$. Sia $\gamma$ una curva chiusa in $D$ (percorsa in verso antiorario) che circonda tutte le $z_1,\dots,z_k$ (e nessun'altra singolarità). Allora:
> $$\oint_\gamma f(z)\,dz = 2\pi i\sum_{j=1}^k\operatorname{Res}(f,z_j)$$

**Idea della dimostrazione**: si "scava" attorno a ciascuna singolarità un piccolo cerchio $\gamma_j$, e si collega $\gamma$ a ciascun $\gamma_j$ con dei "tagli" sottili; la regione risultante (senza le singolarità) è semplicemente connessa e $f$ vi è olomorfa, quindi per il teorema di Cauchy (notebook 9.9) l'integrale lungo il bordo di questa regione (che include $\gamma$ percorsa in un verso e ciascun $\gamma_j$ nel verso opposto, con i tagli che si cancellano a coppie) è zero — da cui $\oint_\gamma f\,dz=\sum_j\oint_{\gamma_j}f\,dz=\sum_j2\pi i\operatorname{Res}(f,z_j)$ (per il notebook 9.13, applicato a ciascun piccolo cerchio).

### Applicazione al calcolo di integrali reali

Il teorema dei residui permette di calcolare integrali reali impropri (notebook 5.18) della forma $\int_{-\infty}^{+\infty}f(x)\,dx$, quando $f$ estesa al piano complesso ha singolarità isolate. La strategia standard:

1. Si considera un contorno $\gamma_R$ formato dal segmento reale $[-R,R]$ più un semicerchio nel semipiano superiore (di raggio $R$).
2. Si calcola $\oint_{\gamma_R}f(z)\,dz$ con il teorema dei residui, sommando i residui **solo** delle singolarità nel semipiano superiore (le uniche racchiuse dal contorno).
3. Si dimostra (spesso con la stima del modulo, notebook 9.8) che il contributo del semicerchio tende a $0$ per $R\to\infty$.
4. Quello che resta, nel limite, è esattamente $\int_{-\infty}^{+\infty}f(x)\,dx$.

### Esempio guidato: $\displaystyle\int_{-\infty}^{+\infty}\frac1{x^2+1}\,dx$

Consideriamo $f(z)=\dfrac1{z^2+1}$, con poli semplici in $z=\pm i$ (fattorizzando come nell'esercizio 4 del notebook 9.13). Solo $z=i$ sta nel semipiano superiore. Il residuo (con la formula $g(z_0)/h'(z_0)$, notebook 9.13, $g=1,h=z^2+1,h'=2z$): $\operatorname{Res}(f,i)=\dfrac1{2i}=-\dfrac i2$.

Per il teorema dei residui, l'integrale lungo il contorno chiuso $\gamma_R$ (segmento + semicerchio) vale $2\pi i\times\left(-\dfrac i2\right)=\pi$. Si dimostra che il contributo del semicerchio tende a $0$ (usando che $|f(z)|\sim1/R^2$ sul semicerchio, mentre la sua lunghezza cresce solo come $R$: il prodotto tende a $0$). Quindi:

$$\int_{-\infty}^{+\infty}\frac{dx}{x^2+1} = \pi$$

— un risultato che, nel Modulo 5, avremmo ottenuto riconoscendo la primitiva $\arctan x$ (notebook 3.16) e calcolando $\arctan(+\infty)-\arctan(-\infty)=\dfrac\pi2-\left(-\dfrac\pi2\right)=\pi$: una **conferma incrociata** soddisfacente, ma qui ottenuta con un metodo che si estende a integrali dove **non esiste** una primitiva elementare.

````python
import sympy as sp

z = sp.symbols('z')
x = sp.symbols('x', real=True)

# Verifica del teorema dei residui: integrale di 1/(x^2+1) su tutto R
f = 1 / (z**2 + 1)
residuo_i = sp.residue(f, z, sp.I)
print("Residuo in z=i:", residuo_i)

integrale_via_residui = 2 * sp.pi * sp.I * residuo_i
print("Integrale (via teorema dei residui):", integrale_via_residui)

# Confronto con il calcolo diretto (Modulo 5, primitiva arctan)
integrale_diretto = sp.integrate(1/(x**2+1), (x, -sp.oo, sp.oo))
print("Integrale diretto (con arctan):", integrale_diretto)

# Un secondo esempio: integrale con SymPy che usa i residui internamente per casi complessi
integrale2 = sp.integrate(1/(x**2+4), (x, -sp.oo, sp.oo))
print("\nIntegrale di 1/(x^2+4):", integrale2)

f2 = 1/(z**2+4)
residuo2 = sp.residue(f2, z, 2*sp.I)
print("Verifica via residuo in z=2i:", sp.simplify(2*sp.pi*sp.I*residuo2))
````

## Esercizio svolto e commentato

**Problema:** usa il teorema dei residui per calcolare $\displaystyle\oint_\gamma\dfrac1{z(z-1)(z+2)}\,dz$, dove $\gamma$ è la circonferenza $|z|=3$ (che circonda tutte e tre le singolarità: $0,1,-2$).

**Ragionamento:**

**Passo 1 — identifichiamo le singolarità**: poli semplici in $z=0$, $z=1$, $z=-2$ (tutti dentro $|z|=3$, essendo $|0|=0<3$, $|1|=1<3$, $|-2|=2<3$).

**Passo 2 — calcoliamo il residuo in $z=0$**, con $g(z)=\dfrac1{(z-1)(z+2)}$ (la funzione senza il fattore che si annulla in $0$) valutata in $z=0$: $\operatorname{Res}(f,0)=\dfrac1{(0-1)(0+2)}=\dfrac1{-2}=-\dfrac12$.

**Passo 3 — calcoliamo il residuo in $z=1$**, con $g(z)=\dfrac1{z(z+2)}$: $\operatorname{Res}(f,1)=\dfrac1{1\times3}=\dfrac13$.

**Passo 4 — calcoliamo il residuo in $z=-2$**, con $g(z)=\dfrac1{z(z-1)}$: $\operatorname{Res}(f,-2)=\dfrac1{(-2)(-3)}=\dfrac16$.

**Passo 5 — sommiamo i residui**: $-\dfrac12+\dfrac13+\dfrac16=\dfrac{-3+2+1}6=\dfrac06=0$.

**Passo 6 — applichiamo il teorema dei residui**: $\oint_\gamma f\,dz=2\pi i\times0=0$.

**Conclusione:** l'integrale vale $0$.

**Perché la somma dei residui che si annulla esattamente non è un caso fortuito, ma riflette una proprietà generale delle funzioni razionali "sufficientemente decrescenti all'infinito":** quando una funzione razionale $f(z)=P(z)/Q(z)$ ha il grado del denominatore **almeno due unità superiore** a quello del numeratore (qui: denominatore di grado 3, numeratore di grado 0, differenza $3$), si dimostra — con un argomento simile a quello usato nell'esempio della teoria per il contributo del semicerchio che tende a zero — che la **somma di tutti i residui** su $\mathbb C$ (inclusi eventuali residui "all'infinito", un concetto avanzato che non trattiamo nei dettagli) deve essere zero; questo esempio, con tutte le singolarità racchiuse dal contorno, ne è una manifestazione diretta, non una coincidenza numerica isolata.

````python
import sympy as sp

z = sp.symbols('z')
f = 1 / (z * (z - 1) * (z + 2))

r0 = sp.residue(f, z, 0)
r1 = sp.residue(f, z, 1)
r2 = sp.residue(f, z, -2)
print("Residui:", r0, r1, r2)
print("Somma:", r0 + r1 + r2)
print("Integrale (2*pi*i*somma):", 2 * sp.pi * sp.I * (r0 + r1 + r2))
````

Somma $=-1+1=0$. Integrale $=2\pi i\times0=0$.

## Riepilogo finale

- Il **teorema dei residui** $\oint_\gamma f\,dz=2\pi i\sum_j\operatorname{Res}(f,z_j)$ somma i contributi di **tutte** le singolarità racchiuse dal contorno — la generalizzazione del fatto chiave del notebook 9.13 a più singolarità.
- L'applicazione più potente: calcolare **integrali reali impropri** (notebook 5.18) che nessuna primitiva elementare risolverebbe, tramite contorni che includono un tratto reale e un arco che tende a "sparire" all'infinito.
- La somma dei residui di una funzione razionale con denominatore di grado sufficientemente maggiore del numeratore tende a zero, un fatto verificabile direttamente negli esempi.
- La formula integrale di Cauchy (notebook 9.10) è un caso particolare del teorema dei residui.
