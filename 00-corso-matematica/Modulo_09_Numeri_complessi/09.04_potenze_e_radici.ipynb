# 9.4 — Potenze e radici

**Prerequisiti:** 9.3 Formula di Eulero, 1.10 Potenze, 0.17 Quantificatori ∀ e ∃.


## Perché ci interessa

In $\mathbb R$, l'equazione $x^2=4$ ha due soluzioni ($\pm2$), ma $x^3=8$ ne ha una sola ($2$: le altre due "si nascondono" nei complessi, come vedremo). In $\mathbb C$, grazie alla forma esponenziale (notebook 9.3), possiamo finalmente rispondere in modo completo: **ogni** numero complesso non nullo ha esattamente $n$ radici $n$-esime distinte, disposte con perfetta simmetria su una circonferenza. Questo notebook sviluppa gli strumenti — la formula di De Moivre e la formula delle radici $n$-esime — che rendono questo calcolo sistematico.

## Teoria, passo dopo passo

### La formula di De Moivre

Per $z=re^{i\theta}$, elevare alla potenza $n$-esima (intero positivo) usa direttamente le proprietà delle potenze (notebook 1.11, 9.3): $z^n=(re^{i\theta})^n=r^ne^{in\theta}$, cioè, in forma trigonometrica:

$$z^n = r^n(\cos(n\theta)+i\sin(n\theta))$$

nota come **formula di De Moivre**. Nel caso particolare $r=1$: $(\cos\theta+i\sin\theta)^n=\cos(n\theta)+i\sin(n\theta)$ — una formula che permette di calcolare potenze di angoli multipli partendo da $\cos\theta,\sin\theta$ (come già intravisto nell'esercizio 4 del notebook 9.3).

### Le radici $n$-esime: quante sono?

Cerchiamo tutte le soluzioni $w\in\mathbb C$ dell'equazione $w^n=z$, con $z=re^{i\theta}$ dato e $n$ intero positivo. Scriviamo $w=\rho e^{i\varphi}$ (incognite $\rho,\varphi$). Per De Moivre: $w^n=\rho^ne^{in\varphi}=re^{i\theta}$. Uguagliando moduli e argomenti:

$$\rho^n = r \quad\Longrightarrow\quad \rho = \sqrt[n]r \qquad\qquad n\varphi = \theta+2k\pi \quad\Longrightarrow\quad \varphi = \frac{\theta+2k\pi}n$$

(l'argomento $\theta$ non è unico, notebook 9.2: differisce per multipli di $2\pi$, da cui il termine $2k\pi$, con $k$ intero **arbitrario**, notebook 0.17 — quantificando su "esiste un intero $k$"). Al variare di $k=0,1,2,\dots,n-1$, si ottengono $n$ valori **distinti** di $\varphi$ (per $k=n$, l'angolo $\varphi$ aumenta di esattamente $2\pi$, tornando allo stesso punto — quindi basta considerare $n$ valori consecutivi di $k$).

### Formula delle radici $n$-esime

$$w_k = \sqrt[n]r\left(\cos\frac{\theta+2k\pi}n+i\sin\frac{\theta+2k\pi}n\right) \qquad k=0,1,\dots,n-1$$

**Ogni** numero complesso non nullo ha **esattamente $n$** radici $n$-esime distinte, tutte con lo stesso modulo $\sqrt[n]r$ (quindi giacciono su una circonferenza di raggio $\sqrt[n]r$) e con argomenti equispaziati di $2\pi/n$: geometricamente, i vertici di un **poligono regolare** di $n$ lati inscritto in quella circonferenza.

### Esempio: le radici quarte dell'unità

Cerchiamo $w^4=1$. Qui $z=1=1\times e^{i0}$ ($r=1,\theta=0$). $w_k=\cos\dfrac{2k\pi}4+i\sin\dfrac{2k\pi}4$ per $k=0,1,2,3$: $w_0=1$, $w_1=i$, $w_2=-1$, $w_3=-i$ — i quattro vertici di un quadrato inscritto nella circonferenza unitaria, che ritroviamo essere esattamente le potenze di $i$ già incontrate nel notebook 9.3!

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Radici n-esime di z = -8 (le radici cubiche, un esempio con z reale negativo)
n_val = 3
z = -8
r = abs(z)
theta = sp.pi  # argomento di -8 (numero reale negativo)

radici = []
for k in range(n_val):
    rho = r**(sp.Rational(1, n_val))
    phi = (theta + 2*k*sp.pi) / n_val
    w = rho * (sp.cos(phi) + sp.I * sp.sin(phi))
    radici.append(sp.simplify(sp.expand(w, complex=True)))

print(f"Radici cubiche di {z}:")
for w in radici:
    print(" ", w, " -> w^3 =", sp.simplify(sp.expand(w**3, complex=True)))

# Confronto con sp.solve
x = sp.symbols('x')
soluzioni_dirette = sp.solve(sp.Eq(x**3, z), x)
print("\nSoluzioni dirette (sp.solve):", soluzioni_dirette)

# Visualizzazione delle radici quarte dell'unità
radici_unita = [np.exp(2j*np.pi*k/4) for k in range(4)]
fig, ax = plt.subplots(figsize=(6, 6))
theta_cerchio = np.linspace(0, 2*np.pi, 100)
ax.plot(np.cos(theta_cerchio), np.sin(theta_cerchio), '--', color='gray', alpha=0.5)
for w in radici_unita:
    ax.plot(w.real, w.imag, 'o', markersize=10)
    ax.annotate(f'{w.real:.0f}+{w.imag:.0f}i', (w.real, w.imag), textcoords="offset points", xytext=(10,5))
ax.set_aspect('equal')
ax.set_title("Le 4 radici quarte dell'unità: vertici di un quadrato")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova tutte le radici cubiche di $z=8i$.

**Ragionamento:**

**Passo 1 — scriviamo $z$ in forma polare.** $|z|=8$. $z=8i$ sta sul semiasse immaginario positivo: $\theta=\pi/2$.

**Passo 2 — applichiamo la formula delle radici**, con $n=3$: $\rho=\sqrt[3]8=2$. $\varphi_k=\dfrac{\pi/2+2k\pi}3$ per $k=0,1,2$.

**Passo 3 — calcoliamo $\varphi_0$** (per $k=0$): $\varphi_0=\dfrac{\pi/2}3=\dfrac\pi6$. $w_0=2\left(\cos\dfrac\pi6+i\sin\dfrac\pi6\right)=2\left(\dfrac{\sqrt3}2+i\dfrac12\right)=\sqrt3+i$.

**Passo 4 — calcoliamo $\varphi_1$** (per $k=1$): $\varphi_1=\dfrac{\pi/2+2\pi}3=\dfrac{5\pi/2}3=\dfrac{5\pi}6$. $w_1=2\left(\cos\dfrac{5\pi}6+i\sin\dfrac{5\pi}6\right)=2\left(-\dfrac{\sqrt3}2+i\dfrac12\right)=-\sqrt3+i$.

**Passo 5 — calcoliamo $\varphi_2$** (per $k=2$): $\varphi_2=\dfrac{\pi/2+4\pi}3=\dfrac{9\pi/2}3=\dfrac{3\pi}2$. $w_2=2\left(\cos\dfrac{3\pi}2+i\sin\dfrac{3\pi}2\right)=2(0-i)=-2i$.

**Conclusione:** le tre radici cubiche di $8i$ sono $\sqrt3+i$, $-\sqrt3+i$, $-2i$.

**Verifica** (su $w_2=-2i$, la più semplice): $w_2^3=(-2i)^3=-8i^3=-8\times(-i)=8i=z$. ✓

**Perché le tre radici sono equispaziate di $\dfrac{2\pi}3=120°$ (verificabile: $\varphi_1-\varphi_0=\frac{5\pi}6-\frac\pi6=\frac{4\pi}6=\frac{2\pi}3$), formando un triangolo equilatero:** questo segue direttamente dalla formula generale $\varphi_k=\frac{\theta+2k\pi}n$, dove la differenza tra $\varphi_{k+1}$ e $\varphi_k$ è sempre $\frac{2\pi}n$, indipendentemente da $\theta$ o da $z$ — una simmetria perfetta che vale per **ogni** numero complesso, non solo per questo esempio: le $n$ radici $n$-esime di qualsiasi numero non nullo formano sempre i vertici di un poligono regolare di $n$ lati.

````python
import sympy as sp

z = 8 * sp.I
r = sp.Abs(z)
theta = sp.arg(z)
n_val = 3

radici = []
for k in range(n_val):
    rho = r**(sp.Rational(1, n_val))
    phi = (theta + 2*k*sp.pi) / n_val
    w = rho * (sp.cos(phi) + sp.I * sp.sin(phi))
    radici.append(sp.simplify(sp.expand(w, complex=True)))

for w in radici:
    print(w, " -> w^3 =", sp.simplify(sp.expand(w**3, complex=True)))
````

## Esercizi da risolvere

2. Trova le radici quadrate di $z=-4$ (dovrebbero essere le familiari $\pm2i$).

4. *(più impegnativo)* Dimostra, in generale, che la somma di tutte le $n$ radici $n$-esime dell'unità è $0$ per ogni $n\ge2$ (suggerimento: le radici sono $\omega^k$ per $k=0,\dots,n-1$, con $\omega=e^{2\pi i/n}$; la loro somma è una serie geometrica, notebook 4.8, di ragione $\omega\ne1$).

**1.** $1+i=\sqrt2\,e^{i\pi/4}$. $(1+i)^8=(\sqrt2)^8e^{i8\pi/4}=16\,e^{i2\pi}=16\times1=16$.

**4.** La somma è $S=\sum_{k=0}^{n-1}\omega^k=1+\omega+\omega^2+\cdots+\omega^{n-1}$, una serie geometrica finita di ragione $\omega$ (notebook 4.8), con formula $S=\dfrac{\omega^n-1}{\omega-1}$ (per $\omega\ne1$, cioè per $n\ge2$). Ma $\omega^n=(e^{2\pi i/n})^n=e^{2\pi i}=1$ (per De Moivre, o direttamente dalla formula di Eulero con $\theta=2\pi$): quindi $S=\dfrac{1-1}{\omega-1}=\dfrac0{\omega-1}=0$.

## Riepilogo finale

- La **formula di De Moivre** $z^n=r^n(\cos n\theta+i\sin n\theta)$ rende immediato elevare a potenza in forma polare.
- **Ogni** numero complesso non nullo ha esattamente $n$ **radici $n$-esime** distinte, con modulo $\sqrt[n]r$ e argomenti equispaziati di $\dfrac{2\pi}n$: i vertici di un poligono regolare.
- Le radici $n$-esime dell'unità, e più in generale di ogni numero complesso, hanno somma $0$ (per $n\ge2$) — conseguenza della formula della serie geometrica.
- In $\mathbb C$, ogni equazione $w^n=z$ (con $z\ne0$) ha sempre soluzioni: una completezza algebrica assente in $\mathbb R$.
