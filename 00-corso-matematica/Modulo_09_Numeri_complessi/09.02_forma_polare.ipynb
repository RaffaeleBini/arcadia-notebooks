# 9.2 — Forma polare

**Prerequisiti:** 9.1 Forma cartesiana, 2.13 Coordinate polari, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Nel notebook 9.1 abbiamo moltiplicato numeri complessi in forma cartesiana: un calcolo algebrico corretto ma che non rivela nulla sul **significato geometrico** dell'operazione. Esattamente come nel notebook 2.13 le coordinate polari semplificavano problemi con simmetria radiale, per i numeri complessi la **forma polare** — modulo e angolo, invece di parte reale e immaginaria — trasforma la moltiplicazione in un'operazione geometricamente trasparente: ruotare e scalare. È il primo passo verso la comprensione profonda della struttura di $\mathbb C$.

## Teoria, passo dopo passo

### Modulo e argomento

Per $z=a+bi\ne0$, oltre al modulo $r=|z|=\sqrt{a^2+b^2}$ (notebook 9.1), definiamo l'**argomento** $\theta=\arg(z)$: l'angolo (in radianti) che il vettore $(a,b)$ forma con il semiasse reale positivo, misurato in senso antiorario — esattamente l'angolo polare del notebook 2.13. Dalle relazioni tra coordinate cartesiane e polari: $a=r\cos\theta$, $b=r\sin\theta$, quindi:

$$z = a+bi = r\cos\theta+ir\sin\theta = r(\cos\theta+i\sin\theta)$$

Questa è la **forma polare** (o trigonometrica) di $z$. L'argomento non è unico: $\theta$ e $\theta+2k\pi$ (per ogni intero $k$) danno lo stesso punto (essendo seno e coseno periodici, notebook 2.12); si sceglie spesso l'**argomento principale** $\theta\in(-\pi,\pi]$.

### Calcolo dell'argomento

Dato $z=a+bi$, si ha $\tan\theta=b/a$ (per $a\ne0$), ma bisogna prestare attenzione al **quadrante** (la funzione arcotangente, notebook 3.16, restituisce valori solo in $(-\pi/2,\pi/2)$, quindi va corretta se $z$ sta nel secondo o terzo quadrante). In pratica, si usa la funzione $\operatorname{atan2}(b,a)$, che tiene conto automaticamente del quadrante.

### Il prodotto in forma polare: ruotare e scalare

Dati $z_1=r_1(\cos\theta_1+i\sin\theta_1)$ e $z_2=r_2(\cos\theta_2+i\sin\theta_2)$, calcoliamo il prodotto:

$$z_1z_2 = r_1r_2\left[(\cos\theta_1\cos\theta_2-\sin\theta_1\sin\theta_2)+i(\sin\theta_1\cos\theta_2+\cos\theta_1\sin\theta_2)\right]$$

Riconosciamo le **formule di addizione** del seno e del coseno (identità trigonometriche, notebook 2.10): $\cos\theta_1\cos\theta_2-\sin\theta_1\sin\theta_2=\cos(\theta_1+\theta_2)$ e $\sin\theta_1\cos\theta_2+\cos\theta_1\sin\theta_2=\sin(\theta_1+\theta_2)$. Quindi:

$$z_1z_2 = r_1r_2\left[\cos(\theta_1+\theta_2)+i\sin(\theta_1+\theta_2)\right]$$

**I moduli si moltiplicano, gli argomenti si sommano.** Geometricamente: moltiplicare per $z_2$ significa **ruotare** di un angolo $\theta_2$ e **scalare** di un fattore $r_2$ — esattamente la trasformazione lineare del notebook 6.14 (matrice di rotazione composta con una scalatura)! Questo spiega perché la moltiplicazione per $i$ (che ha $r=1,\theta=\pi/2$) corrisponde a una rotazione di $90°$: $i\times1=i$, $i\times i=i^2=-1$ (rotazione di $180°$ totale), coerente.

### La divisione in forma polare

Analogamente (o invertendo la formula del prodotto): $\dfrac{z_1}{z_2}=\dfrac{r_1}{r_2}\left[\cos(\theta_1-\theta_2)+i\sin(\theta_1-\theta_2)\right]$ — i moduli si dividono, gli argomenti si sottraggono.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Conversione cartesiana <-> polare
z = 1 + sp.sqrt(3) * sp.I
r = sp.Abs(z)
theta = sp.arg(z)
print("z =", z, " -> r =", r, " theta =", theta, "(=", sp.deg(theta), "gradi)")

# Verifica: prodotto in forma polare
z1 = 2 * (sp.cos(sp.pi/6) + sp.I * sp.sin(sp.pi/6))
z2 = 3 * (sp.cos(sp.pi/3) + sp.I * sp.sin(sp.pi/3))
prodotto = sp.expand(z1 * z2)
print("\nz1*z2 (calcolo diretto):", sp.simplify(prodotto))

r1r2 = 2 * 3
theta_somma = sp.pi/6 + sp.pi/3
prodotto_polare = r1r2 * (sp.cos(theta_somma) + sp.I * sp.sin(theta_somma))
print("z1*z2 (formula polare, r1*r2 e theta1+theta2):", sp.simplify(prodotto_polare))

# Visualizzazione: moltiplicare per i ruota di 90 gradi
fig, ax = plt.subplots(figsize=(6, 6))
punti = [1, 1j, 1j**2, 1j**3]
colori = ['blue', 'green', 'red', 'purple']
for p, c in zip(punti, colori):
    ax.annotate('', xy=(p.real, p.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle='->', color=c, lw=2))
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)
ax.set_title("1, i, i², i³: ogni moltiplicazione per i ruota di 90°")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** scrivi $z=-4+4i$ in forma polare, poi usa la forma polare per calcolare $z^2$ (senza sviluppare il quadrato del binomio).

**Ragionamento:**

**Passo 1 — calcoliamo il modulo**: $r=\sqrt{(-4)^2+4^2}=\sqrt{32}=4\sqrt2$.

**Passo 2 — individuiamo il quadrante.** $z=-4+4i$ ha parte reale negativa e parte immaginaria positiva: secondo quadrante.

**Passo 3 — calcoliamo l'argomento.** $\tan\theta=\dfrac4{-4}=-1$. L'arcotangente di $-1$ dà $-\pi/4$, ma questo è nel quarto quadrante, non nel secondo: dobbiamo **correggere** aggiungendo $\pi$ (per portarci nel semipiano corretto, notebook 3.16): $\theta=-\dfrac\pi4+\pi=\dfrac{3\pi}4$.

**Passo 4 — verifichiamo**: $\cos(3\pi/4)=-\dfrac{\sqrt2}2$, $\sin(3\pi/4)=\dfrac{\sqrt2}2$ (angoli notevoli, notebook 2.9). $r\cos\theta=4\sqrt2\times\left(-\dfrac{\sqrt2}2\right)=-4$. ✓ $r\sin\theta=4\sqrt2\times\dfrac{\sqrt2}2=4$. ✓

**Passo 5 — calcoliamo $z^2$ usando la regola "moduli si moltiplicano, argomenti si sommano"** (con $z_1=z_2=z$): $|z^2|=r\times r=r^2=(4\sqrt2)^2=32$. $\arg(z^2)=\theta+\theta=2\theta=\dfrac{3\pi}2$.

**Passo 6 — riconvertiamo in forma cartesiana**: $\cos(3\pi/2)=0$, $\sin(3\pi/2)=-1$. $z^2=32(0+i\times(-1))=-32i$.

**Conclusione:** $z^2=-32i$.

**Verifica diretta** (per controllo): $z^2=(-4+4i)^2=16-32i+16i^2=16-32i-16=-32i$. ✓ Coincide.

**Perché la forma polare rende il calcolo delle potenze molto più semplice del prodotto notevole in forma cartesiana, specialmente per potenze alte:** elevare al quadrato in forma cartesiana richiede sviluppare un binomio (un calcolo che si complica rapidamente per potenze superiori, richiedendo il binomio di Newton); in forma polare, elevare alla $n$-esima potenza significa semplicemente **elevare il modulo alla $n$ ed moltiplicare l'argomento per $n$** — una regola che, come vedremo nel prossimo notebook (formula di De Moivre), rende immediato calcolare anche $z^{100}$, un calcolo proibitivo in forma cartesiana.

````python
import sympy as sp

z = -4 + 4*sp.I
r = sp.Abs(z)
theta = sp.arg(z)
print("r =", r, " theta =", theta)

z_polare = r * (sp.cos(theta) + sp.I * sp.sin(theta))
print("Verifica forma polare:", sp.simplify(z_polare))

z2_polare = r**2 * (sp.cos(2*theta) + sp.I * sp.sin(2*theta))
print("\nz^2 (via polare):", sp.simplify(z2_polare))
print("z^2 (diretto):", sp.expand(z**2))
````

## Riepilogo finale

- La **forma polare** $z=r(\cos\theta+i\sin\theta)$ usa modulo $r=|z|$ e argomento $\theta=\arg(z)$ (l'angolo polare del notebook 2.13) invece di parte reale e immaginaria.
- Il **prodotto** in forma polare moltiplica i moduli e **somma** gli argomenti: geometricamente, una rotazione composta con una scalatura.
- La **divisione** divide i moduli e sottrae gli argomenti.
- Calcolare l'argomento richiede attenzione al quadrante: l'arcotangente da sola non basta sempre.
