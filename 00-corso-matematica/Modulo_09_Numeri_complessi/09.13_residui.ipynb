# 9.13 — Residui

**Prerequisiti:** 9.12 Singolarità, 9.8 Integrali complessi, 5.9 Teoremi fondamentali della derivazione.


## Perché ci interessa

Nel notebook 9.8 abbiamo calcolato $\oint_{|z|=1}\dfrac1z\,dz=2\pi i$, un caso isolato. Nel notebook 9.12 abbiamo osservato che, per **qualsiasi** funzione con uno sviluppo di Laurent, solo il coefficiente $c_{-1}$ (il residuo) contribuisce all'integrale lungo un cerchio attorno alla singolarità. Questo notebook formalizza questa osservazione e sviluppa **formule pratiche** per calcolare i residui — specialmente nel caso dei poli — senza dover scrivere l'intera serie di Laurent ogni volta: uno strumento di calcolo che preparerà il terreno per il teorema dei residui.

## Teoria, passo dopo passo

### Definizione di residuo

Se $z_0$ è una singolarità isolata di $f$, con sviluppo di Laurent $f(z)=\sum_nc_n(z-z_0)^n$ nella corona bucata attorno a $z_0$, il **residuo** di $f$ in $z_0$ è il coefficiente:

$$\operatorname{Res}(f,z_0) := c_{-1}$$

### Perché solo $c_{-1}$ conta nell'integrazione

> **Fatto chiave.** Per ogni cerchio $\gamma$ centrato in $z_0$ (contenuto nella corona dove vale lo sviluppo): $\oint_\gamma(z-z_0)^n\,dz=\begin{cases}2\pi i&\text{se }n=-1\\0&\text{se }n\ne-1\end{cases}$

(il caso $n=-1$ è esattamente il calcolo del notebook 9.8; per $n\ne-1$, $(z-z_0)^n$ ha una primitiva $\dfrac{(z-z_0)^{n+1}}{n+1}$ ben definita — anche per $n<-1$, purché $n\ne-1$ — quindi l'integrale lungo una curva chiusa si annulla per il corollario del teorema di Cauchy, notebook 9.9, applicato localmente). Integrando termine per termine la serie di Laurent (un'operazione lecita per serie che convergono uniformemente sul cerchio, un dettaglio tecnico che qui diamo per buono):

$$\oint_\gamma f(z)\,dz = \sum_nc_n\oint_\gamma(z-z_0)^n\,dz = c_{-1}\times2\pi i = 2\pi i\operatorname{Res}(f,z_0)$$

**Ogni** altro termine della serie contribuisce zero: solo il residuo "sopravvive" all'integrazione.

### Calcolo del residuo per un polo semplice (ordine 1)

Se $z_0$ è un polo **semplice** (ordine 1), $f(z)=\dfrac{c_{-1}}{z-z_0}+c_0+c_1(z-z_0)+\cdots$. Moltiplicando per $(z-z_0)$ e prendendo il limite $z\to z_0$: $(z-z_0)f(z)\to c_{-1}$ (gli altri termini, moltiplicati per $(z-z_0)\to0$, si annullano). Quindi:

$$\operatorname{Res}(f,z_0) = \lim_{z\to z_0}(z-z_0)f(z)$$

**Caso particolare utile**: se $f(z)=\dfrac{g(z)}{h(z)}$ con $g(z_0)\ne0$ e $h$ ha uno zero semplice in $z_0$ (cioè $h(z_0)=0$, $h'(z_0)\ne0$), allora $(z-z_0)f(z)=g(z)\dfrac{z-z_0}{h(z)}\to g(z_0)\times\dfrac1{h'(z_0)}$ (riconoscendo il reciproco della definizione di derivata, notebook 5.8, applicata a $h$ in $z_0$). Quindi:

$$\operatorname{Res}(f,z_0) = \frac{g(z_0)}{h'(z_0)}$$

una formula estremamente pratica.

### Calcolo del residuo per un polo di ordine $m$

Se $z_0$ è un polo di ordine $m$, si dimostra (moltiplicando per $(z-z_0)^m$ per eliminare la parte principale, poi derivando $m-1$ volte per isolare il coefficiente cercato, e infine passando al limite):

$$\operatorname{Res}(f,z_0) = \frac1{(m-1)!}\lim_{z\to z_0}\frac{d^{m-1}}{dz^{m-1}}\left[(z-z_0)^mf(z)\right]$$

Per singolarità **eliminabili**, il residuo è ovviamente $0$ (non c'è parte principale). Per singolarità **essenziali**, in generale bisogna calcolare esplicitamente la serie di Laurent (notebook 9.11), non esistendo formule scorciatoia.

````python
import sympy as sp

z = sp.symbols('z')

# Residuo di f(z) = e^z / z^2 in z=0 (polo di ordine 2, dal notebook 9.11-9.12)
f = sp.exp(z) / z**2
m = 2
residuo = sp.limit(sp.diff((z - 0)**m * f, z, m - 1), z, 0) / sp.factorial(m - 1)
print("Residuo di e^z/z^2 in z=0 (formula generale):", residuo)

# Confronto: direttamente dalla serie di Laurent (coefficiente di 1/z)
laurent = sp.series(f, z, 0, 3)
print("Serie di Laurent:", laurent, "(coefficiente di 1/z:", laurent.coeff(z, -1), ")")

# Residuo tramite g(z0)/h'(z0), per f(z) = z / (z^2+1) in z0 = i (polo semplice)
g = z
h = z**2 + 1
z0 = sp.I
residuo2 = g.subs(z, z0) / sp.diff(h, z).subs(z, z0)
print("\nResiduo di z/(z^2+1) in z=i:", sp.simplify(residuo2))

# Confronto diretto: limite di (z-i)*f(z)
f2 = z / (z**2 + 1)
residuo2_diretto = sp.limit((z - sp.I) * f2, z, sp.I)
print("Confronto (limite diretto):", residuo2_diretto)
````

## Esercizio svolto e commentato

**Problema:** calcola il residuo di $f(z)=\dfrac1{z^2(z-2)}$ in ciascuna delle sue singolarità.

**Ragionamento:**

**Passo 1 — identifichiamo le singolarità.** Il denominatore si annulla in $z=0$ (con molteplicità 2, dal fattore $z^2$) e in $z=2$ (molteplicità 1).

**Passo 2 — classifichiamo (dal notebook 9.12, stesso tipo di analisi dell'esercizio 4 di quel notebook)**: $z=0$ è un polo di ordine $2$; $z=2$ è un polo semplice.

**Passo 3 — calcoliamo il residuo in $z=2$ (polo semplice)**, con la formula $g(z_0)/h'(z_0)$: scriviamo $f(z)=\dfrac{g(z)}{h(z)}$ con $g(z)=\dfrac1{z^2}$ (la parte "regolare" vicino a $z=2$) e $h(z)=z-2$, $h'(z)=1$. $\operatorname{Res}(f,2)=\dfrac{g(2)}{h'(2)}=\dfrac{1/4}1=\dfrac14$.

**Passo 4 — calcoliamo il residuo in $z=0$ (polo di ordine $m=2$)**, con la formula generale: $\operatorname{Res}(f,0)=\dfrac1{1!}\lim_{z\to0}\dfrac d{dz}\left[z^2\times\dfrac1{z^2(z-2)}\right]=\lim_{z\to0}\dfrac d{dz}\left[\dfrac1{z-2}\right]$.

**Passo 5 — calcoliamo la derivata**: $\dfrac d{dz}\left[(z-2)^{-1}\right]=-(z-2)^{-2}$ (regola della potenza, notebook 5.9, trasferita al caso complesso, notebook 9.6).

**Passo 6 — valutiamo il limite** (qui, banalmente, sostituendo $z=0$, essendo la funzione risultante continua in $0$): $-\dfrac1{(0-2)^2}=-\dfrac14$.

**Conclusione:** $\operatorname{Res}(f,0)=-\dfrac14$, $\operatorname{Res}(f,2)=\dfrac14$.

**Perché conviene sempre verificare l'ordine esatto del polo (passo 2) prima di scegliere quale formula applicare:** usare la formula del polo semplice ($\lim(z-z_0)f(z)$) su un polo di ordine 2 darebbe un limite **infinito** (non eliminerebbe completamente la parte principale), un errore facile da individuare ma che richiede comunque di rifare il calcolo con la formula corretta; classificare l'ordine con precisione (come fatto sistematicamente nel notebook 9.12) è quindi un passo preliminare indispensabile, non solo una formalità, perché determina **quale** delle due formule (o quale derivata, per ordini superiori) sia effettivamente applicabile.

````python
import sympy as sp

z = sp.symbols('z')
f = 1 / (z**2 * (z - 2))

residuo_in_2 = sp.residue(f, z, 2)
residuo_in_0 = sp.residue(f, z, 0)
print("Residuo in z=2:", residuo_in_2)
print("Residuo in z=0:", residuo_in_0)
````

## Riepilogo finale

- Il **residuo** $\operatorname{Res}(f,z_0)=c_{-1}$ è il coefficiente della serie di Laurent che, per il fatto chiave dimostrato in questo notebook, è l'**unico** a sopravvivere nell'integrazione lungo un cerchio: $\oint_\gamma f\,dz=2\pi i\operatorname{Res}(f,z_0)$.
- Per un **polo semplice**: $\operatorname{Res}(f,z_0)=\lim_{z\to z_0}(z-z_0)f(z)$, o $g(z_0)/h'(z_0)$ per $f=g/h$.
- Per un **polo di ordine $m$**: una formula con derivata $(m-1)$-esima, che richiede di aver classificato correttamente l'ordine (notebook 9.12) prima di applicarla.
- Le singolarità eliminabili hanno residuo $0$; le singolarità essenziali richiedono in genere la serie di Laurent completa.
