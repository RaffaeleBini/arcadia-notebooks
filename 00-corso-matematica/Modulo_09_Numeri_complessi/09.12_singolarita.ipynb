# 9.12 — Singolarità

**Prerequisiti:** 9.11 Serie di Laurent, 5.6 Continuità, 5.4 Limiti.


## Perché ci interessa

Nel notebook 9.11 abbiamo visto esempi diversi di serie di Laurent: alcune con parte principale finita (come $\dfrac1{z^2}$, un solo termine negativo), altre — vedremo — con infiniti termini negativi. Questa differenza non è un dettaglio tecnico: riflette **tipi qualitativamente diversi** di comportamento della funzione vicino al punto dove non è olomorfa. Classificare le **singolarità** in tre categorie nette — eliminabili, poli, essenziali — è il passo che precede il calcolo dei residui (notebook 9.13), rendendolo sistematico invece che caso per caso.

## Teoria, passo dopo passo

### Cos'è una singolarità

$z_0$ è un **punto singolare isolato** di $f$ se $f$ è olomorfa in una corona bucata $0<|z-z_0|<R$ ma **non** è definita (o non è olomorfa) esattamente in $z_0$. In quella corona, $f$ ammette uno sviluppo di Laurent (notebook 9.11): $f(z)=\sum_{n=-\infty}^\infty c_n(z-z_0)^n$. La classificazione dipende **interamente** dalla parte principale (i coefficienti $c_n$ con $n<0$).

### Tipo 1: singolarità eliminabile

Se **tutti** i coefficienti $c_n$ con $n<0$ sono nulli (parte principale assente), $z_0$ è una **singolarità eliminabile**: la serie di Laurent è in realtà una serie di Taylor ordinaria, e $f$ si **estende** a una funzione olomorfa anche in $z_0$, ponendo $f(z_0):=c_0$.

**Esempio**: $f(z)=\dfrac{\sin z}z$ (non definita in $z=0$). Dalla serie $\sin z=z-\frac{z^3}{3!}+\cdots$: $f(z)=1-\dfrac{z^2}{3!}+\dfrac{z^4}{5!}-\cdots$, senza potenze negative. Ponendo $f(0)=1$ (coerente col limite notevole, notebook 5.5), $f$ diventa olomorfa ovunque — la stessa funzione già incontrata nell'esercizio svolto del notebook 9.10.

**Criterio pratico**: $z_0$ è una singolarità eliminabile se e solo se $\lim_{z\to z_0}f(z)$ esiste **finito** (un fatto analogo alla continuità del notebook 5.6, ora in versione complessa).

### Tipo 2: polo

Se la parte principale ha un numero **finito** e non nullo di termini, cioè $c_{-m}\ne0$ per qualche $m>0$ ma $c_n=0$ per $n<-m$, $z_0$ è un **polo di ordine $m$**. La serie ha la forma $f(z)=\dfrac{c_{-m}}{(z-z_0)^m}+\cdots+\dfrac{c_{-1}}{z-z_0}+c_0+c_1(z-z_0)+\cdots$

**Esempio**: $f(z)=\dfrac{e^z}{z^2}$ (notebook 9.11) ha parte principale $\dfrac1{z^2}+\dfrac1z$: un **polo di ordine 2** in $z=0$ (il termine di ordine più negativo è $z^{-2}$).

**Criterio pratico**: $z_0$ è un polo di ordine $m$ se e solo se $\lim_{z\to z_0}(z-z_0)^mf(z)$ esiste **finito e non nullo**. Equivalentemente, $|f(z)|\to+\infty$ per $z\to z_0$ (a differenza del caso eliminabile, dove $f$ resta limitata).

### Tipo 3: singolarità essenziale

Se la parte principale ha **infiniti** termini non nulli, $z_0$ è una **singolarità essenziale**.

**Esempio**: $f(z)=e^{1/z}$. Sostituendo $w=1/z$ nella serie di $e^w$ (notebook 9.10): $e^{1/z}=1+\dfrac1z+\dfrac1{2!z^2}+\dfrac1{3!z^3}+\cdots$ — infiniti termini con potenze negative di $z$: singolarità essenziale in $z=0$.

**Comportamento (teorema di Casorati-Weierstrass, enunciato senza dimostrazione)**: vicino a una singolarità essenziale, $f$ assume valori che si avvicinano a **ogni** numero complesso possibile (tranne al più uno) — un comportamento estremamente "selvaggio", ben diverso dal semplice "esplodere all'infinito" di un polo.

### Perché questa classificazione è la chiave per i residui

Il coefficiente $c_{-1}$ della parte principale — il **residuo** — gioca un ruolo speciale (lo vedremo nel notebook 9.13): è l'unico coefficiente che sopravvive nell'integrazione lungo un cerchio attorno a $z_0$ (ricordando che $\oint\dfrac1{z-z_0}dz=2\pi i$, mentre $\oint(z-z_0)^ndz=0$ per ogni altro intero $n\ne-1$, un fatto verificabile con lo stesso calcolo del notebook 9.8). Sapere se una singolarità è un polo (dove il residuo si calcola con formule algebriche dirette) o essenziale (dove serve l'intera serie di Laurent) determina la strategia di calcolo.

````python
import sympy as sp

z = sp.symbols('z')

# Classificazione automatica con SymPy (tramite l'ordine del polo, se esiste)
funzioni = {
    "sin(z)/z (eliminabile)": sp.sin(z)/z,
    "e^z/z^2 (polo ordine 2)": sp.exp(z)/z**2,
    "e^(1/z) (essenziale)": sp.exp(1/z),
}

for nome, f in funzioni.items():
    laurent = sp.series(f, z, 0, 4)
    print(f"{nome}: {laurent}")

# Verifica del criterio del limite per la singolarità eliminabile
limite_eliminabile = sp.limit(sp.sin(z)/z, z, 0)
print("\nLimite di sin(z)/z per z->0:", limite_eliminabile, "(finito: eliminabile)")

# Verifica del criterio per il polo: lim (z-z0)^m * f(z)
limite_polo = sp.limit(z**2 * (sp.exp(z)/z**2), z, 0)
print("Limite di z^2 * e^z/z^2 per z->0:", limite_polo, "(finito e non nullo: polo di ordine 2)")
````

## Esercizio svolto e commentato

**Problema:** classifica la singolarità di $f(z)=\dfrac{z^2-1}{z-1}$ in $z=1$.

**Ragionamento:**

**Passo 1 — verifichiamo che $z=1$ è effettivamente un punto dove $f$ non è definita**: il denominatore si annulla, $1-1=0$.

**Passo 2 — semplifichiamo l'espressione**, riconoscendo una differenza di quadrati al numeratore (notebook 1.2): $z^2-1=(z-1)(z+1)$. Quindi $f(z)=\dfrac{(z-1)(z+1)}{z-1}=z+1$, per $z\ne1$ (semplificazione lecita, essendo $z-1\ne0$ nel dominio dove stiamo lavorando).

**Passo 3 — riconosciamo che $f(z)=z+1$ (per $z\ne1$) è già un polinomio**: la sua "serie di Laurent" attorno a $z_0=1$ è semplicemente $f(z)=(z-1)+2$ (riscrivendo $z+1=(z-1)+2$), **senza alcun termine di potenza negativa**.

**Passo 4 — applichiamo il criterio del limite**: $\lim_{z\to1}f(z)=\lim_{z\to1}(z+1)=2$, un limite **finito**.

**Conclusione:** $z=1$ è una **singolarità eliminabile**: ponendo $f(1):=2$, la funzione si estende a un polinomio (quindi intera) su tutto $\mathbb C$.

**Perché questo esempio, apparentemente "artificiale", è in realtà il prototipo di un fenomeno molto comune:** ogni volta che una funzione razionale ha, al numeratore e al denominatore, un **fattore comune** che si cancella (qui, $z-1$), il punto dove il denominatore originale si annullava diventa automaticamente una singolarità **eliminabile**, non un vero polo — perché, dopo la semplificazione algebrica, la funzione risultante è ovunque ben definita in quel punto. Questo è esattamente il motivo per cui il **criterio del limite** (passo 4) è così importante: senza semplificare esplicitamente (un procedimento a volte complicato, in casi meno immediati di questo), il limite rivela comunque immediatamente se la singolarità è "reale" (un polo, dove il limite è infinito) o solo "apparente" (eliminabile, dove il limite esiste finito).

````python
import sympy as sp

z = sp.symbols('z')
f = (z**2 - 1) / (z - 1)

f_semplificato = sp.simplify(f)
print("f semplificata:", f_semplificato)

limite = sp.limit(f, z, 1)
print("Limite per z->1:", limite, "(finito: singolarità eliminabile)")
````

**1.** $\lim_{z\to0}z^3\times\dfrac1{z^3}=\lim_{z\to0}1=1$: finito e non nullo, con $m=3$: **polo di ordine 3**.

**3.** $\sin(1/z)=\dfrac1z-\dfrac1{3!z^3}+\dfrac1{5!z^5}-\cdots$: infiniti termini con potenze negative: **singolarità essenziale**.

**4.** In $z=0$: $\lim_{z\to0}z^2\times\dfrac1{z^2(z-1)}=\lim_{z\to0}\dfrac1{z-1}=\dfrac1{-1}=-1$, finito e non nullo: **polo di ordine 2**. In $z=1$: $\lim_{z\to1}(z-1)\times\dfrac1{z^2(z-1)}=\lim_{z\to1}\dfrac1{z^2}=1$, finito e non nullo: **polo di ordine 1** (detto anche **polo semplice**). Le due singolarità della stessa funzione hanno quindi **ordine diverso**: 2 in $z=0$, 1 in $z=1$.

## Riepilogo finale

- Una **singolarità isolata** $z_0$ si classifica in base alla parte principale della serie di Laurent: **eliminabile** (nessun termine negativo, limite finito), **polo di ordine $m$** (finiti termini negativi, $\lim(z-z_0)^mf(z)$ finito e non nullo), **essenziale** (infiniti termini negativi).
- Le singolarità eliminabili sono "false" singolarità: la funzione si estende a olomorfa anche lì.
- I poli fanno esplodere $|f(z)|\to\infty$; le singolarità essenziali hanno un comportamento estremamente irregolare (teorema di Casorati-Weierstrass).
- Il coefficiente $c_{-1}$ (il **residuo**) gioca un ruolo speciale nell'integrazione, indipendentemente dal tipo di singolarità — l'argomento del prossimo notebook.
