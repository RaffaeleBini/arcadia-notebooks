# 9.8 — Integrali complessi

**Prerequisiti:** 9.7 Derivabilità complessa, 7.18 Integrali curvilinei.


## Perché ci interessa

Nel notebook 7.18 abbiamo integrato campi vettoriali lungo curve nel piano. Un integrale di una funzione complessa lungo una curva nel piano complesso è concettualmente **la stessa costruzione**, ma la struttura algebrica di $\mathbb C$ (in particolare, il fatto che $f(z)\,dz$ è un prodotto tra numeri complessi, non un prodotto scalare) produce risultati sorprendenti — a partire dal prossimo notebook, dove scopriremo che l'integrale di una funzione **olomorfa** lungo una curva chiusa è quasi sempre zero, un fatto con conseguenze profonde.

## Teoria, passo dopo passo

### Curve nel piano complesso

Una curva $\gamma:[a,b]\to\mathbb C$ è data da $\gamma(t)=x(t)+iy(t)$ — esattamente una curva parametrica del notebook 7.18, con le due componenti reali unite in un unico numero complesso. La sua derivata $\gamma'(t)=x'(t)+iy'(t)$ è il vettore tangente, come prima.

### Definizione dell'integrale complesso

Dato $f:D\to\mathbb C$ continua e $\gamma:[a,b]\to D$ una curva, l'**integrale complesso** di $f$ lungo $\gamma$ è:

$$\int_\gamma f(z)\,dz := \int_a^bf(\gamma(t))\,\gamma'(t)\,dt$$

dove l'integrale a destra è un integrale di una funzione complessa di variabile reale $t$: si calcola integrando separatamente parte reale e parte immaginaria (entrambe funzioni reali di $t$), come già fatto implicitamente nel notebook 8.11 con le trasformate.

### Collegamento con gli integrali curvilinei reali

Scrivendo $f(z)=u+iv$ e $dz=dx+i\,dy$ (formalmente, da $\gamma'(t)dt=(x'(t)+iy'(t))dt$), il prodotto $f\,dz$ si sviluppa:

$$f(z)\,dz = (u+iv)(dx+i\,dy) = (u\,dx-v\,dy)+i(v\,dx+u\,dy)$$

quindi:

$$\int_\gamma f(z)\,dz = \int_\gamma(u\,dx-v\,dy) + i\int_\gamma(v\,dx+u\,dy)$$

Riconosciamo **due integrali curvilinei reali** (notebook 7.18), del tipo $\int_\gamma P\,dx+Q\,dy$! L'integrale complesso è quindi "due integrali curvilinei in uno", combinati secondo la struttura algebrica di $\mathbb C$.

### Proprietà fondamentali

Come per gli integrali curvilinei reali: **linearità** ($\int_\gamma(af+bg)\,dz=a\int_\gamma f\,dz+b\int_\gamma g\,dz$), e la **stima del modulo** (l'analogo complesso della disuguaglianza triangolare per gli integrali):

$$\left|\int_\gamma f(z)\,dz\right| \le \int_a^b|f(\gamma(t))|\,|\gamma'(t)|\,dt \le \max_{z\in\gamma}|f(z)|\times L(\gamma)$$

dove $L(\gamma)$ è la lunghezza della curva (notebook 7.18) — una stima spesso usata per dimostrare che certi integrali tendono a zero.

### Un esempio fondamentale: $\displaystyle\oint_{|z|=1}\frac1z\,dz$

Calcoliamo l'integrale di $f(z)=1/z$ lungo la circonferenza unitaria $\gamma(t)=e^{it}$, $t\in[0,2\pi]$ (percorsa in senso antiorario, notebook 7.23). $\gamma'(t)=ie^{it}$. Quindi:

$$\oint_{|z|=1}\frac1z\,dz = \int_0^{2\pi}\frac1{e^{it}}\times ie^{it}\,dt = \int_0^{2\pi}i\,dt = i\times2\pi = 2\pi i$$

**Un risultato non nullo!** Nonostante $f(z)=1/z$ sia olomorfa ovunque tranne che nell'origine (dove non è definita), l'integrale lungo un cerchio che "circonda" quel punto singolare non si annulla — un fatto che sarà la chiave di volta del teorema dei residui (notebook 9.14).

````python
import sympy as sp

t = sp.symbols('t', real=True)

# Integrale di 1/z lungo la circonferenza unitaria
gamma = sp.exp(sp.I * t)
gamma_prime = sp.diff(gamma, t)
f_su_gamma = 1 / gamma

integranda = sp.simplify(f_su_gamma * gamma_prime)
print("Integranda f(gamma(t))*gamma'(t):", integranda)

integrale = sp.integrate(integranda, (t, 0, 2*sp.pi))
print("Integrale di 1/z lungo |z|=1:", integrale)

# Confronto: integrale di z lungo la stessa circonferenza (funzione intera, senza singolarità)
f2_su_gamma = gamma
integranda2 = sp.simplify(f2_su_gamma * gamma_prime)
integrale2 = sp.integrate(integranda2, (t, 0, 2*sp.pi))
print("\nIntegrale di z lungo |z|=1:", integrale2, "(funzione olomorfa ovunque: integrale nullo)")
````

## Esercizio svolto e commentato

**Problema:** calcola $\displaystyle\int_\gamma z^2\,dz$, dove $\gamma(t)=t+it^2$, $t\in[0,1]$ (un arco di parabola nel piano complesso, dal punto $0$ al punto $1+i$).

**Ragionamento:**

**Passo 1 — calcoliamo $\gamma'(t)$**: $\gamma'(t)=1+2it$ (derivando componente per componente, notebook 7.18).

**Passo 2 — valutiamo $f(\gamma(t))=\gamma(t)^2$** (quadrato di binomio, notebook 1.2): $(t+it^2)^2=t^2+2t\times it^2+(it^2)^2=t^2+2it^3+i^2t^4=t^2+2it^3-t^4$.

**Passo 3 — moltiplichiamo per $\gamma'(t)$**: $(t^2-t^4+2it^3)(1+2it)$. Sviluppiamo: $t^2-t^4+2it^3+2it^3-2it^5+4i^2t^4=t^2-t^4+4it^3-2it^5-4t^4=t^2-5t^4+4it^3-2it^5$ (raccogliendo parte reale e immaginaria, usando $i^2=-1$).

**Passo 4 — integriamo termine per termine**, da $0$ a $1$: parte reale, $\int_0^1(t^2-5t^4)dt=\left[\dfrac{t^3}3-t^5\right]_0^1=\dfrac13-1=-\dfrac23$. Parte immaginaria: $\int_0^1(4t^3-2t^5)dt=\left[t^4-\dfrac{t^6}3\right]_0^1=1-\dfrac13=\dfrac23$.

**Conclusione:** $\displaystyle\int_\gamma z^2\,dz=-\dfrac23+\dfrac23i$.

**Verifica con una scorciatoia (anticipando il prossimo notebook):** poiché $f(z)=z^2$ è olomorfa (funzione intera), con primitiva $F(z)=z^3/3$ (esattamente come nel caso reale, $F'(z)=z^2$ per le regole trasferite dal notebook 9.6), vale un analogo del teorema fondamentale del calcolo (notebook 5.16): $\int_\gamma z^2dz=F(\gamma(1))-F(\gamma(0))=F(1+i)-F(0)=\dfrac{(1+i)^3}3-0$. Calcolando $(1+i)^3=(1+i)(1+i)^2=(1+i)(2i)=2i+2i^2=2i-2=-2+2i$: $\dfrac{-2+2i}3=-\dfrac23+\dfrac23i$. ✓ Coincide esattamente!

**Perché questa coincidenza non è un caso, ma un'anteprima di un risultato generale:** per funzioni olomorfe con primitiva esplicita, l'integrale lungo una curva dipende **solo dai punti estremi**, non dalla forma della curva — esattamente come per i campi conservativi (notebook 7.20). Questo fatto, che qui abbiamo solo intravisto con un calcolo diretto, sarà formalizzato rigorosamente nel prossimo notebook con il teorema di Cauchy.

````python
import sympy as sp

t = sp.symbols('t', real=True)
gamma = t + sp.I * t**2
gamma_prime = sp.diff(gamma, t)

integranda = sp.expand(gamma**2 * gamma_prime, complex=True)
print("Integranda:", integranda)

integrale = sp.integrate(integranda, (t, 0, 1))
print("Integrale:", sp.simplify(integrale))

# Verifica con la primitiva
F = lambda z: z**3 / 3
verifica = F(1 + sp.I) - F(0)
print("Verifica con F(gamma(1))-F(gamma(0)):", sp.expand(verifica, complex=True))
````

## Riepilogo finale

- L'**integrale complesso** $\int_\gamma f(z)\,dz=\int_a^bf(\gamma(t))\gamma'(t)\,dt$ generalizza l'integrale curvilineo (notebook 7.18) alla struttura algebrica di $\mathbb C$: si decompone in due integrali curvilinei reali.
- L'integrale di $1/z$ lungo un cerchio attorno all'origine vale $2\pi i$: **non nullo**, nonostante $1/z$ sia olomorfa quasi ovunque — un'anteprima della teoria dei residui.
- Per funzioni **olomorfe con primitiva esplicita**, l'integrale dipende solo dagli estremi della curva (come i campi conservativi, notebook 7.20); per funzioni **non olomorfe** come $\bar z$, dipende dal percorso.
- La **stima del modulo** $\left|\int_\gamma f\,dz\right|\le\max|f|\times L(\gamma)$ è uno strumento pratico per limitare gli integrali senza calcolarli esplicitamente.
