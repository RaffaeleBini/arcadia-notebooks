## Teoria, passo dopo passo

### L'unità immaginaria

Definiamo un nuovo simbolo $i$ (l'**unità immaginaria**) con la proprietà:

$$i^2 = -1$$

Con questo singolo assioma aggiuntivo, costruiamo l'insieme $\mathbb C$ dei **numeri complessi**: espressioni della forma $z=a+bi$, con $a,b\in\mathbb R$. Il numero $a$ si chiama **parte reale** ($\operatorname{Re}(z)=a$), e $b$ **parte immaginaria** ($\operatorname{Im}(z)=b$). Se $b=0$, $z=a$ è semplicemente un numero reale: $\mathbb R\subset\mathbb C$, esattamente come $\mathbb Q\subset\mathbb R$.

### Operazioni: somma e prodotto

La somma si fa componente per componente, come per i vettori di $\mathbb R^2$ (notebook 6.1): $(a+bi)+(c+di)=(a+c)+(b+d)i$. Il **prodotto**, invece, è la vera novità: si moltiplica come un binomio (prodotto notevole, notebook 1.2), usando $i^2=-1$ per semplificare:

$$(a+bi)(c+di) = ac+adi+bci+bdi^2 = (ac-bd)+(ad+bc)i$$

**Esempio**: $(2+3i)(1-i)=2-2i+3i-3i^2=2+i-3(-1)=2+i+3=5+i$.

### $\mathbb C$ come piano: la forma cartesiana

Un numero complesso $z=a+bi$ si rappresenta geometricamente come il punto (o vettore) $(a,b)$ nel **piano complesso** (o piano di Argand-Gauss): l'asse orizzontale è l'asse reale, quello verticale l'asse immaginario — esattamente il piano cartesiano del notebook 2.1, con la somma di numeri complessi che coincide con la somma di vettori (notebook 6.1). Questa rappresentazione si chiama **forma cartesiana** o **algebrica**.

### Il coniugato e il modulo

Il **coniugato** di $z=a+bi$ è $\bar z=a-bi$ (riflessione rispetto all'asse reale). Una proprietà cruciale: $z\bar z=(a+bi)(a-bi)=a^2-abi+abi-b^2i^2=a^2+b^2$ (differenza di quadrati, notebook 1.2, applicata con $i^2=-1$) — sempre un numero **reale non negativo**. Questo definisce il **modulo**:

$$|z| = \sqrt{z\bar z} = \sqrt{a^2+b^2}$$

esattamente la norma euclidea del vettore $(a,b)$ (notebook 6.2, 7.2): il modulo di $z$ è la sua distanza dall'origine nel piano complesso.

### Divisione: razionalizzare con il coniugato

Per dividere, si moltiplica numeratore e denominatore per il coniugato del denominatore (una tecnica analoga alla razionalizzazione con le radici, notebook 0.6):

$$\frac1{c+di} = \frac{c-di}{(c+di)(c-di)} = \frac{c-di}{c^2+d^2}$$

trasformando il denominatore in un numero reale, con cui si può dividere normalmente.

### $\mathbb C$ come spazio vettoriale, e qualcosa in più

Come insieme con somma e prodotto per uno scalare reale, $\mathbb C$ è uno spazio vettoriale su $\mathbb R$ di dimensione 2 (notebook 6.3), con base $\{1,i\}$ — identico a $\mathbb R^2$. Ma $\mathbb C$ ha **in più** un prodotto **tra elementi** (non solo per scalari), che $\mathbb R^2$ da solo non possiede: questa struttura aggiuntiva (un **campo**, che generalizza le proprietà algebriche di $\mathbb Q$ e $\mathbb R$ viste nel notebook 0.5) è ciò che rende $\mathbb C$ molto più ricco della semplice coppia di coordinate.

````python
import sympy as sp

I = sp.I  # unità immaginaria di SymPy

z1 = 2 + 3*I
z2 = 1 - I

print("z1 =", z1, " z2 =", z2)
print("z1 + z2 =", z1 + z2)
print("z1 * z2 =", sp.expand(z1 * z2))

# Coniugato e modulo
print("\nconiugato di z1:", sp.conjugate(z1))
print("z1 * conjugate(z1) =", sp.expand(z1 * sp.conjugate(z1)))
print("|z1| =", sp.Abs(z1))
print("|z1|^2 =", sp.Abs(z1)**2)

# Divisione
z3 = z1 / z2
print("\nz1/z2 =", sp.simplify(z3))

# Verifica: risolvere x^2+1=0 in C
x = sp.symbols('x')
soluzioni = sp.solve(sp.Eq(x**2 + 1, 0), x)
print("\nSoluzioni di x^2+1=0:", soluzioni)
````

## Esercizio svolto e commentato

**Problema:** risolvi $x^2-4x+13=0$ in $\mathbb C$.

**Ragionamento:**

**Passo 1 — applichiamo la formula risolutiva** (notebook 1.6): $x=\dfrac{4\pm\sqrt{16-52}}2=\dfrac{4\pm\sqrt{-36}}2$.

**Passo 2 — riscriviamo la radice del numero negativo** usando $i$: $\sqrt{-36}=\sqrt{36\times(-1)}=\sqrt{36}\times\sqrt{-1}=6i$ (estraendo il fattore reale $36$ e isolando $\sqrt{-1}=i$, per definizione).

**Passo 3 — sostituiamo**: $x=\dfrac{4\pm6i}2=2\pm3i$.

**Conclusione:** le due soluzioni sono $x_1=2+3i$ e $x_2=2-3i$ — **coniugate** l'una dell'altra.

**Verifica** (sostituendo $x_1=2+3i$): $x_1^2=(2+3i)^2=4+12i+9i^2=4+12i-9=-5+12i$ (quadrato di binomio, notebook 1.2). $-4x_1=-8-12i$. Sommando: $x_1^2-4x_1+13=(-5+12i)+(-8-12i)+13=(-5-8+13)+(12-12)i=0+0i=0$. ✓

**Perché le due soluzioni sono sempre coniugate, quando i coefficienti dell'equazione sono reali:** la formula risolutiva produce $x=\dfrac{-b\pm\sqrt{b^2-4ac}}{2a}$; se $b^2-4ac<0$, la radice $\sqrt{b^2-4ac}=i\sqrt{4ac-b^2}$ è puramente immaginaria, e le due soluzioni differiscono **solo** nel segno di questa parte immaginaria — cioè sono l'una il coniugato dell'altra. Questo fatto (che vale per **ogni** equazione a coefficienti reali, non solo di secondo grado, come vedremo più avanti nel corso) riflette una simmetria profonda: coniugare un'equazione a coefficienti reali non la cambia, quindi se $z$ è soluzione anche $\bar z$ lo è.

````python
import sympy as sp

x = sp.symbols('x')
soluzioni = sp.solve(sp.Eq(x**2 - 4*x + 13, 0), x)
print("Soluzioni:", soluzioni)

x1 = soluzioni[0]
verifica = sp.expand(x1**2 - 4*x1 + 13)
print("Verifica (deve fare 0):", verifica)
````

## Riepilogo finale

- I **numeri complessi** $\mathbb C=\{a+bi:a,b\in\mathbb R\}$ estendono $\mathbb R$ con l'unità immaginaria $i$, tale che $i^2=-1$, colmando la lacuna delle equazioni senza soluzioni reali.
- Somma e prodotto si calcolano algebricamente, usando $i^2=-1$; il **coniugato** $\bar z=a-bi$ e il **modulo** $|z|=\sqrt{a^2+b^2}=\sqrt{z\bar z}$ sono strumenti fondamentali.
- Geometricamente, $\mathbb C$ si identifica con il piano cartesiano (notebook 2.1); la somma coincide con la somma vettoriale (notebook 6.1).
- Ogni equazione a coefficienti reali con discriminante negativo ha soluzioni complesse **coniugate** — un fatto che spiega retroattivamente le radici complesse coniugate incontrate nel notebook 8.4.
