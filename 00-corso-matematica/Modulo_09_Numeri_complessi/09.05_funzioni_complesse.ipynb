**Prerequisiti:** 9.4 Potenze e radici, 3.1 Concetto di funzione, 7.4 Funzioni di più variabili.


## Perché ci interessa

Finora abbiamo studiato singoli numeri complessi. Il passo successivo, naturale quanto in ogni altro modulo di questo corso, è studiare **funzioni** $f:\mathbb C\to\mathbb C$: regole che associano a ogni numero complesso un altro numero complesso. Ma c'è una difficoltà nuova rispetto al notebook 3.1: il grafico di $f$ vivrebbe in $\mathbb C\times\mathbb C$, uno spazio a **quattro** dimensioni reali (due per il dominio, due per il codominio, notebook 7.4) — impossibile da visualizzare direttamente. In questo notebook sviluppiamo le tecniche per capire comunque il comportamento di una funzione complessa, prima di affrontare (nei prossimi notebook) la nozione cruciale di derivabilità complessa.

## Teoria, passo dopo passo

### Funzione complessa come trasformazione del piano

Una funzione $f:\mathbb C\to\mathbb C$ associa a ogni $z=x+iy$ un valore $f(z)=u+iv$. Possiamo pensare $f$ come una **trasformazione** che manda punti del piano $(x,y)$ in punti del piano $(u,v)$ — esattamente un caso particolare di funzione vettoriale $\mathbb R^2\to\mathbb R^2$ (notebook 7.4), con la peculiarità che dominio e codominio hanno anche la struttura algebrica di $\mathbb C$ (notebook 9.1).

### Parte reale e immaginaria come funzioni di due variabili

Scrivendo $z=x+iy$ e $f(z)=u(x,y)+iv(x,y)$, ogni funzione complessa si decompone in **due funzioni reali di due variabili reali**, $u$ e $v$ (esattamente il tipo di funzione studiato nel Modulo 7). Per esempio, per $f(z)=z^2$: $f(x+iy)=(x+iy)^2=x^2-y^2+2xyi$, quindi $u(x,y)=x^2-y^2$ e $v(x,y)=2xy$.

### Polinomi ed esponenziale complessi

I **polinomi** $f(z)=a_nz^n+\cdots+a_1z+a_0$ (con coefficienti complessi) si calcolano esattamente come nel caso reale (notebook 1.1), usando $i^2=-1$. L'**esponenziale complesso** si definisce combinando la formula di Eulero (notebook 9.3) con la proprietà delle potenze: per $z=x+iy$,

$$e^z = e^{x+iy} = e^x\,e^{iy} = e^x(\cos y+i\sin y)$$

Questa definizione **estende** l'esponenziale reale (per $y=0$, si ritrova $e^x$) e preserva la proprietà fondamentale $e^{z_1+z_2}=e^{z_1}e^{z_2}$ (verificabile dalla definizione, usando le proprietà delle potenze reali e la formula di addizione di seno e coseno, come già visto per il prodotto in forma polare, notebook 9.2).

### Visualizzare funzioni complesse: la griglia trasformata

Un modo efficace di "vedere" una funzione complessa: disegnare una griglia nel dominio (rette verticali e orizzontali) e tracciarne le **immagini** tramite $f$. Per $f(z)=z^2$, per esempio, le rette verticali $x=c$ (costante) si trasformano in parabole nel piano $(u,v)$ (poiché $u=x^2-y^2=c^2-y^2$ e $v=2cy$, quindi $y=v/(2c)$ e $u=c^2-v^2/(4c^2)$: una parabola in $v$). Questo tipo di visualizzazione rivela la geometria della trasformazione, altrimenti nascosta.

### Periodicità dell'esponenziale complesso

Una proprietà sorprendente, assente nel caso reale: $e^{z+2\pi i}=e^z\,e^{2\pi i}=e^z\times1=e^z$ (usando $e^{2\pi i}=\cos(2\pi)+i\sin(2\pi)=1$, formula di Eulero). L'esponenziale complesso è **periodico** di periodo $2\pi i$ (lungo la direzione immaginaria) — un fenomeno che non ha alcun analogo nell'esponenziale reale, dove $e^x$ è strettamente crescente e mai periodica.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y', real=True)
z = x + sp.I * y

f = z**2
f_espanso = sp.expand(f)
u = sp.re(f_espanso)
v = sp.im(f_espanso)
print("f(z) = z^2 =", f_espanso)
print("u(x,y) =", u, " v(x,y) =", v)

# Verifica della periodicita' dell'esponenziale complesso
z_sym = sp.symbols('z')
periodo = sp.simplify(sp.exp(z_sym + 2*sp.pi*sp.I) - sp.exp(z_sym))
print("\ne^(z+2*pi*i) - e^z =", periodo)

# Visualizzazione: griglia trasformata da f(z) = z^2
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Griglia originale
for c in np.linspace(-2, 2, 9):
    ys = np.linspace(-2, 2, 100)
    axes[0].plot(c*np.ones_like(ys), ys, color='blue', alpha=0.4)
    xs = np.linspace(-2, 2, 100)
    axes[0].plot(xs, c*np.ones_like(xs), color='red', alpha=0.4)
axes[0].set_title("Griglia nel dominio (piano z)")
axes[0].set_aspect('equal')

# Griglia trasformata da f(z)=z^2
for c in np.linspace(-2, 2, 9):
    ys = np.linspace(-2, 2, 100)
    us = c**2 - ys**2
    vs = 2*c*ys
    axes[1].plot(us, vs, color='blue', alpha=0.4)
    xs = np.linspace(-2, 2, 100)
    us2 = xs**2 - c**2
    vs2 = 2*xs*c
    axes[1].plot(us2, vs2, color='red', alpha=0.4)
axes[1].set_title("Immagine tramite f(z)=z² (piano w)")
axes[1].set_xlim(-4, 4)
axes[1].set_ylim(-4, 4)
axes[1].set_aspect('equal')
plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova $u(x,y)$ e $v(x,y)$ per $f(z)=e^z$, e verifica che $|f(z)|=e^x$ (dipende solo dalla parte reale di $z$).

**Ragionamento:**

**Passo 1 — applichiamo direttamente la definizione** dell'esponenziale complesso (teoria): $f(x+iy)=e^{x+iy}=e^x(\cos y+i\sin y)=e^x\cos y+ie^x\sin y$.

**Passo 2 — identifichiamo parte reale e immaginaria**: $u(x,y)=e^x\cos y$, $v(x,y)=e^x\sin y$.

**Passo 3 — calcoliamo il modulo** $|f(z)|=\sqrt{u^2+v^2}$: $u^2+v^2=e^{2x}\cos^2y+e^{2x}\sin^2y=e^{2x}(\cos^2y+\sin^2y)=e^{2x}\times1=e^{2x}$ (identità fondamentale, notebook 2.10, raccogliendo $e^{2x}$ a fattor comune).

**Passo 4 — estraiamo la radice**: $|f(z)|=\sqrt{e^{2x}}=e^x$ (essendo $e^x>0$ sempre, la radice è semplicemente $e^x$).

**Conclusione:** $|e^z|=e^x=e^{\operatorname{Re}(z)}$ — il modulo di $e^z$ dipende **solo** dalla parte reale di $z$, non dalla parte immaginaria.

**Perché questo risultato ha senso, ripensando alla forma polare (notebook 9.2):** avevamo scritto $e^z=e^x(\cos y+i\sin y)$, che è **esattamente** la forma polare di un numero complesso con modulo $e^x$ e argomento $y$! Quindi $e^z$ per $z=x+iy$ è il numero complesso di modulo $e^x$ e argomento $y$ — questo spiega immediatamente sia perché $|e^z|=e^x$ (il modulo è, per costruzione, quel fattore $e^x$), sia perché $e^z$ è periodico in $y$ con periodo $2\pi$ (essendo l'argomento definito a meno di multipli di $2\pi$, notebook 9.2): tutte le proprietà di $e^z$ discendono naturalmente dal riconoscere questa struttura, senza bisogno di calcoli aggiuntivi.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = sp.exp(x + sp.I*y)
f_espanso = sp.expand(f, complex=True)
print("e^(x+iy) espanso:", f_espanso)

u = sp.re(f_espanso)
v = sp.im(f_espanso)
print("u =", u, " v =", v)

modulo = sp.sqrt(u**2 + v**2)
print("modulo =", sp.simplify(modulo))
````

## Riepilogo finale

- Una funzione complessa $f(z)=u(x,y)+iv(x,y)$ si decompone in due funzioni reali di due variabili (Modulo 7), ma dominio e codominio hanno anche la struttura algebrica di $\mathbb C$.
- L'**esponenziale complesso** $e^z=e^x(\cos y+i\sin y)$ estende quello reale, preservando $e^{z_1+z_2}=e^{z_1}e^{z_2}$, ma introduce una **periodicità** (di periodo $2\pi i$) assente nel caso reale.
- $|e^z|=e^{\operatorname{Re}(z)}$: il modulo dipende solo dalla parte reale.
- Visualizzare l'immagine di una griglia tramite $f$ rivela la geometria di una trasformazione complessa, altrimenti difficile da intuire.
