# 9.11 — Serie di Laurent

**Prerequisiti:** 9.10 Serie di Taylor (complesse), 4.8 Serie geometrica, 5.19 Serie numeriche (rigorose).


## Perché ci interessa

La serie di Taylor (notebook 9.10) funziona solo dove $f$ è **olomorfa**. Ma funzioni come $\dfrac1z$ o $\dfrac{e^z}{z^2}$ hanno punti dove non sono definite — eppure vorremmo comunque uno sviluppo in serie, utile vicino a quei punti "problematici". La **serie di Laurent** risolve questo: permette anche potenze **negative** di $(z-z_0)$, catturando esattamente il comportamento di una funzione vicino a un punto dove non è olomorfa. È lo strumento che, nei prossimi due notebook, ci permetterà di classificare le singolarità e calcolare i residui — il cuore del teorema dei residui.

## Teoria, passo dopo passo

### La serie di Laurent

Sia $f$ olomorfa in una **corona circolare** $A=\{z: r<|z-z_0|<R\}$ (un anello, non un disco pieno — notebook 7.3, un insieme "bucato" al centro). Allora $f$ si sviluppa in una **serie di Laurent**:

$$f(z) = \sum_{n=-\infty}^{\infty}c_n(z-z_0)^n = \cdots+\frac{c_{-2}}{(z-z_0)^2}+\frac{c_{-1}}{z-z_0}+c_0+c_1(z-z_0)+c_2(z-z_0)^2+\cdots$$

convergente in tutto l'anello $A$. La parte con potenze **negative** ($n<0$) si chiama **parte principale**; la parte con potenze non negative si chiama **parte regolare** (o analitica) — coincide con una serie di Taylor ordinaria.

### I coefficienti: una formula integrale

I coefficienti si calcolano con una formula analoga a quella delle derivate (notebook 9.10), ma **valida anche per potenze negative** (dove non ha senso parlare di "derivata $n$-esima" per $n<0$):

$$c_n = \frac1{2\pi i}\oint_\gamma\frac{f(z)}{(z-z_0)^{n+1}}\,dz$$

dove $\gamma$ è una qualsiasi curva chiusa nell'anello $A$ che circonda $z_0$ una volta. In pratica, i coefficienti si trovano quasi sempre **algebricamente**, riconoscendo serie note (geometrica, esponenziale), senza calcolare integrali direttamente — esattamente come per le serie di Taylor reali (notebook 5.20).

### Esempio guidato: $f(z)=\dfrac1z$ attorno a $z_0=0$

Questa funzione **è già** la sua stessa serie di Laurent (banalmente): $f(z)=z^{-1}$, con $c_{-1}=1$ e tutti gli altri coefficienti nulli.

### Esempio guidato: $f(z)=\dfrac{e^z}{z^2}$ attorno a $z_0=0$

Partiamo dalla serie di Taylor nota di $e^z$ (notebook 9.10): $e^z=1+z+\dfrac{z^2}{2!}+\dfrac{z^3}{3!}+\cdots$. Dividiamo termine per termine per $z^2$:

$$\frac{e^z}{z^2} = \frac1{z^2}+\frac1z+\frac1{2!}+\frac z{3!}+\frac{z^2}{4!}+\cdots$$

Riconosciamo la serie di Laurent: parte principale $\dfrac1{z^2}+\dfrac1z$ (potenze negative), parte regolare $\dfrac1{2!}+\dfrac z{3!}+\cdots$ (potenze non negative). Questo procedimento — dividere una serie di Taylor nota per una potenza di $z$ — è **il metodo pratico standard** per trovare le serie di Laurent, evitando quasi sempre il calcolo diretto dei coefficienti tramite l'integrale.

### Esempio guidato: $f(z)=\dfrac1{z(z-1)}$ attorno a $z_0=0$

Per $|z|<1$: usiamo la scomposizione in fratti semplici (notebook 1.4) $\dfrac1{z(z-1)}=\dfrac{-1}z+\dfrac1{z-1}$. Il secondo termine, $\dfrac1{z-1}=\dfrac{-1}{1-z}$, si sviluppa con la serie geometrica (notebook 4.8, valida per $|z|<1$): $\dfrac{-1}{1-z}=-(1+z+z^2+\cdots)$. Quindi:

$$\frac1{z(z-1)} = \frac{-1}z-1-z-z^2-\cdots$$

una serie di Laurent con parte principale $-1/z$ e parte regolare $-1-z-z^2-\cdots$, valida nell'anello $0<|z|<1$.

````python
import sympy as sp

z = sp.symbols('z')

# Serie di Laurent di e^z/z^2 attorno a z0=0
f1 = sp.exp(z) / z**2
laurent1 = sp.series(f1, z, 0, 5)
print("Serie di Laurent di e^z/z^2:", laurent1)

# Serie di Laurent di 1/(z(z-1)) attorno a z0=0
f2 = 1 / (z * (z - 1))
laurent2 = sp.series(f2, z, 0, 5)
print("\nSerie di Laurent di 1/(z(z-1)):", laurent2)

# Confronto: coefficienti tramite scomposizione manuale
fratti = sp.apart(f2, z)
print("\nScomposizione in fratti semplici:", fratti)
````

## Esercizio svolto e commentato

**Problema:** trova la serie di Laurent di $f(z)=\dfrac{\sin z}{z^3}$ attorno a $z_0=0$, e identifica la parte principale.

**Ragionamento:**

**Passo 1 — partiamo dalla serie di Taylor nota di $\sin z$** (dal notebook 5.14/9.10, estesa a $\mathbb C$): $\sin z=z-\dfrac{z^3}{3!}+\dfrac{z^5}{5!}-\dfrac{z^7}{7!}+\cdots$.

**Passo 2 — dividiamo ogni termine per $z^3$**:

$$\frac{\sin z}{z^3} = \frac z{z^3}-\frac{z^3}{3!\,z^3}+\frac{z^5}{5!\,z^3}-\frac{z^7}{7!\,z^3}+\cdots = \frac1{z^2}-\frac1{3!}+\frac{z^2}{5!}-\frac{z^4}{7!}+\cdots$$

**Passo 3 — identifichiamo parte principale e parte regolare.** La parte principale è $\dfrac1{z^2}$ (l'unico termine con potenza negativa: notiamo che il termine con $z^1/z^3=1/z^2$ è presente, ma **manca** completamente il termine $1/z$, perché lo sviluppo di $\sin z$ non ha termine di grado $2$ — essendo $\sin z$ una funzione **dispari**, notebook 3.8, contiene solo potenze dispari, quindi $\sin z/z^3$ contiene solo potenze **pari** di $z$, comprese quelle negative). La parte regolare è $-\dfrac1{3!}+\dfrac{z^2}{5!}-\dfrac{z^4}{7!}+\cdots=-\dfrac16+\dfrac{z^2}{120}-\cdots$.

**Conclusione:** $\dfrac{\sin z}{z^3}=\dfrac1{z^2}-\dfrac16+\dfrac{z^2}{120}-\cdots$, con parte principale $\dfrac1{z^2}$ (coefficiente $c_{-2}=1$, e $c_{-1}=0$).

**Perché riconoscere la parità di $\sin z$ (passo 3) permette di prevedere in anticipo quali coefficienti saranno nulli, senza calcolarli esplicitamente:** dato che $\sin z$ contiene solo potenze **dispari** di $z$ (dispari, notebook 3.8), dividendo per $z^3$ (dispari) si ottiene una funzione che contiene solo potenze **pari** (dispari meno dispari, nell'esponente, dà pari) — quindi **tutti** i coefficienti di potenza dispari (incluso $c_{-1}$, il coefficiente di $1/z$, cruciale per il calcolo dei residui nel notebook 9.13) sono automaticamente **zero**, senza bisogno di calcolarli uno per uno: la parità della funzione originale si riflette direttamente nella struttura della sua serie di Laurent.

````python
import sympy as sp

z = sp.symbols('z')
f = sp.sin(z) / z**3
laurent = sp.series(f, z, 0, 6)
print("Serie di Laurent di sin(z)/z^3:", laurent)
````

**1.** $f(z)=z^{-3}$ è già la sua serie di Laurent: $c_{-3}=1$, tutti gli altri coefficienti nulli.

**4.** $\dfrac1{z-2}=\dfrac1z\times\dfrac1{1-2/z}=\dfrac1z\left(1+\dfrac2z+\dfrac4{z^2}+\cdots\right)=\dfrac1z+\dfrac2{z^2}+\dfrac4{z^3}+\cdots$ (tutte potenze negative, valida per $|z|>2$). Questo esempio mostra che la stessa funzione $f(z)=1/(z-2)$ ammette **due sviluppi di Laurent diversi**, a seconda della regione considerata (dentro o fuori dal cerchio $|z|=2$) — un fenomeno senza analogo nelle serie di Taylor, dove lo sviluppo attorno a un punto è unico.

## Riepilogo finale

- La **serie di Laurent** $f(z)=\sum_{n=-\infty}^\infty c_n(z-z_0)^n$ estende la serie di Taylor con potenze **negative**, valida in una corona circolare attorno a un punto dove $f$ non è necessariamente olomorfa.
- La **parte principale** (potenze negative) cattura il comportamento singolare; la **parte regolare** (potenze non negative) è una serie di Taylor ordinaria.
- In pratica, si trova dividendo serie di Taylor note o usando la serie geometrica, senza calcolare l'integrale dei coefficienti direttamente.
- La stessa funzione può avere sviluppi di Laurent **diversi** in anelli diversi attorno allo stesso punto — un fenomeno assente nelle serie di Taylor.
