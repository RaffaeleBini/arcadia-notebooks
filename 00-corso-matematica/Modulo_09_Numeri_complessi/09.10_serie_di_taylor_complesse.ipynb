# 9.10 — Serie di Taylor (complesse)

**Prerequisiti:** 9.9 Teorema di Cauchy, 5.14 Formula di Taylor, 5.20 Serie di potenze (rigorose).


## Perché ci interessa

Nel notebook 5.14, una funzione reale derivabile infinite volte **non era garantita** essere uguale alla propria serie di Taylor (un controesempio classico esiste, anche se non lo abbiamo costruito esplicitamente). Nel mondo complesso, succede qualcosa di quasi miracoloso: **ogni** funzione olomorfa è automaticamente uguale alla propria serie di Taylor, in un intero disco attorno a ogni punto. Questo notebook arriva a questo risultato passando per uno strumento fondamentale — la **formula integrale di Cauchy** — che esprime il valore di una funzione olomorfa in un punto interamente in termini del suo comportamento sul bordo di una regione.

## Teoria, passo dopo passo

### La formula integrale di Cauchy

> **Teorema.** Sia $f$ olomorfa su un dominio semplicemente connesso $D$, e $\gamma$ una curva chiusa (percorsa in verso antiorario) in $D$ che circonda un punto $z_0$. Allora:
> $$f(z_0) = \frac1{2\pi i}\oint_\gamma\frac{f(z)}{z-z_0}\,dz$$

**Idea della dimostrazione**: si applica il teorema di Cauchy (notebook 9.9) alla funzione $g(z)=\dfrac{f(z)-f(z_0)}{z-z_0}$ (olomorfa su $D\setminus\{z_0\}$, e "quasi olomorfa" anche in $z_0$, essendo il rapporto incrementale di $f$), deformando $\gamma$ in un piccolo cerchio attorno a $z_0$ (un procedimento standard basato sul teorema di Cauchy applicato alla regione "ad anello" tra $\gamma$ e il piccolo cerchio); sul piccolo cerchio, $g(z)\to f'(z_0)$ per il raggio che tende a $0$, ma il termine dominante nell'integrale rimanente è esattamente l'integrale di $\dfrac1{z-z_0}$ già calcolato nel notebook 9.8 (che vale $2\pi i$), da cui la formula.

Questo risultato è straordinario: il valore di $f$ in **un solo punto interno** $z_0$ è completamente determinato dai valori di $f$ **sul bordo** $\gamma$ — una funzione olomorfa non può "variare liberamente" all'interno, è rigidamente vincolata dal proprio comportamento sul contorno.

### Derivate di ogni ordine, automaticamente

Derivando formalmente la formula di Cauchy rispetto a $z_0$ (sotto il segno di integrale, un'operazione lecita che non giustifichiamo nei dettagli tecnici), si ottiene una formula per le derivate di **ogni ordine**:

$$f^{(n)}(z_0) = \frac{n!}{2\pi i}\oint_\gamma\frac{f(z)}{(z-z_0)^{n+1}}\,dz$$

**Conseguenza sorprendente**: se $f$ è olomorfa (derivabile **una** volta, in senso complesso), questa formula mostra che $f$ è automaticamente derivabile **infinite** volte! A differenza del caso reale (dove derivabilità una volta non implica affatto derivabilità due volte, figuriamoci infinite), l'olomorfia — una condizione apparentemente "debole" — implica immediatamente la massima regolarità possibile.

### Il teorema di Taylor per funzioni olomorfe

> **Teorema.** Se $f$ è olomorfa su un disco $B(z_0,R)$, allora per ogni $z$ in quel disco:
> $$f(z) = \sum_{n=0}^\infty\frac{f^{(n)}(z_0)}{n!}(z-z_0)^n$$
> e la serie **converge** in tutto il disco.

Questo è l'analogo esatto della serie di Taylor del notebook 5.14, ma con una differenza cruciale: nel caso complesso, l'uguaglianza vale **sempre** (non serve verificare che il resto tenda a zero, un controllo delicato nel caso reale) — è una conseguenza diretta della formula integrale di Cauchy applicata alle derivate. Le funzioni olomorfe si dicono per questo anche **analitiche**: i due termini sono, in analisi complessa, sinonimi.

### Esempio: la serie di $e^z$

Dalle formule di derivazione trasferite (notebook 9.6), $(e^z)^{(n)}=e^z$ per ogni $n$; in $z_0=0$: $f^{(n)}(0)=e^0=1$. Quindi $e^z=\sum_{n=0}^\infty\dfrac{z^n}{n!}$ — la **stessa** serie usata come **definizione** di $e^{i\theta}$ nel notebook 9.3, ora giustificata anche come sviluppo di Taylor, per ogni $z\in\mathbb C$ (raggio di convergenza infinito, come nel caso reale, notebook 5.20).

````python
import sympy as sp

z, z0 = sp.symbols('z z0')

# Verifica della formula integrale di Cauchy per f(z)=z^2, z0=1, su un cerchio che circonda z0
t = sp.symbols('t', real=True)
z0_val = 1
raggio = 2
gamma = z0_val + raggio * sp.exp(sp.I * t)  # cerchio che circonda z0=1

f = lambda w: w**2
integranda = f(gamma) / (gamma - z0_val) * sp.diff(gamma, t)
integrale = sp.integrate(sp.expand(integranda, complex=True), (t, 0, 2*sp.pi))
formula_cauchy = integrale / (2 * sp.pi * sp.I)

print("f(z0) tramite formula integrale di Cauchy:", sp.simplify(formula_cauchy))
print("f(z0) diretto:", f(z0_val))

# Serie di Taylor di e^z attorno a 0, confrontata con la serie usata nel notebook 9.3
serie_taylor = sp.series(sp.exp(z), z, 0, 6)
print("\nSerie di Taylor di e^z:", serie_taylor)
````

## Esercizio svolto e commentato

**Problema:** usa la formula integrale di Cauchy per calcolare $\displaystyle\oint_{|z|=2}\dfrac{\sin z}{z}\,dz$.

**Ragionamento:**

**Passo 1 — riscriviamo l'integrale nella forma della formula di Cauchy**: $\dfrac{\sin z}z=\dfrac{f(z)}{z-z_0}$, con $f(z)=\sin z$ (olomorfa su tutto $\mathbb C$, essendo combinazione di esponenziali interi, notebook 9.9) e $z_0=0$.

**Passo 2 — verifichiamo che $z_0=0$ è circondato dal cerchio $|z|=2$**: sì, $|0|=0<2$.

**Passo 3 — applichiamo la formula integrale di Cauchy**: $f(z_0)=\dfrac1{2\pi i}\displaystyle\oint_{|z|=2}\dfrac{f(z)}{z-z_0}dz$, quindi $\displaystyle\oint_{|z|=2}\dfrac{\sin z}zdz=2\pi i\times f(0)$.

**Passo 4 — calcoliamo $f(0)=\sin(0)=0$.**

**Conclusione:** $\displaystyle\oint_{|z|=2}\dfrac{\sin z}z\,dz=2\pi i\times0=0$.

**Perché questo risultato — che l'integrale è zero — poteva essere previsto anche senza la formula di Cauchy, ma la formula lo rende immediato:** la funzione $g(z)=\sin(z)/z$ ha, in realtà, una **singolarità eliminabile** in $z=0$ (concetto che formalizzeremo nel notebook 9.12): il limite $\lim_{z\to0}\frac{\sin z}z=1$ (limite notevole, notebook 5.5, valido anche in senso complesso) esiste finito, quindi $g$ si **estende** a una funzione olomorfa anche in $0$ — e per il teorema di Cauchy (notebook 9.9), l'integrale di una funzione olomorfa lungo una curva chiusa in un dominio dove è ovunque olomorfa è sempre $0$. La formula integrale di Cauchy, applicata qui a $\sin z$ (che **è** genuinamente olomorfa ovunque, incluso in $z=0$), dà lo stesso risultato per una via diversa — un'utile verifica incrociata tra i due strumenti sviluppati in questo e nel notebook precedente.

````python
import sympy as sp

t = sp.symbols('t', real=True)
gamma = 2 * sp.exp(sp.I * t)
gamma_prime = sp.diff(gamma, t)

f_su_gamma = sp.sin(gamma) / gamma
integranda = sp.expand(f_su_gamma * gamma_prime, complex=True)
integrale = sp.integrate(integranda, (t, 0, 2*sp.pi))
print("Integrale di sin(z)/z lungo |z|=2:", sp.simplify(integrale))
````

## Riepilogo finale

- La **formula integrale di Cauchy** esprime $f(z_0)$ come integrale di $f$ sul bordo di una regione che circonda $z_0$: il valore interno è completamente determinato dal comportamento sul bordo.
- Derivando formalmente, si ottengono formule per **tutte** le derivate di $f$: conseguenza, ogni funzione olomorfa è automaticamente derivabile infinite volte.
- **Ogni funzione olomorfa è uguale alla propria serie di Taylor** in ogni disco dove è olomorfa — senza le sottigliezze del caso reale (notebook 5.14): olomorfa e analitica sono sinonimi in $\mathbb C$.
- La serie di $e^z$ (usata come definizione nel notebook 9.3) è ora giustificata anche come sviluppo di Taylor.
