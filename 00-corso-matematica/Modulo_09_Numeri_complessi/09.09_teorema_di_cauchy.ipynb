# 9.9 — Teorema di Cauchy

**Prerequisiti:** 9.8 Integrali complessi, 9.7 Derivabilità complessa, 7.23 Teorema di Green.


## Perché ci interessa

Nel notebook 9.8 abbiamo verificato, con un calcolo diretto, che $\int_\gamma z^2\,dz$ non dipende dal percorso (per una funzione olomorfa con primitiva), mentre $\int_\gamma\bar z\,dz$ (non olomorfa) sì. Il **teorema di Cauchy** eleva questa osservazione a teorema generale: per **ogni** funzione olomorfa, l'integrale lungo una curva chiusa è **sempre zero**. È probabilmente il risultato singolo più importante dell'analisi complessa, da cui discendono — come vedremo nei prossimi notebook — le serie di Taylor automatiche, la formula integrale di Cauchy, e il teorema dei residui.

## Teoria, passo dopo passo

### Enunciato

> **Teorema (di Cauchy, o di Cauchy-Goursat).** Se $f$ è olomorfa su un dominio **semplicemente connesso** $D$ (senza "buchi", notebook 7.23) e $\gamma$ è una curva chiusa contenuta in $D$, allora:
> $$\oint_\gamma f(z)\,dz = 0$$

### Dimostrazione (nel caso di derivate parziali continue, con il teorema di Green)

Usiamo la decomposizione del notebook 9.8: $\oint_\gamma f\,dz=\oint_\gamma(u\,dx-v\,dy)+i\oint_\gamma(v\,dx+u\,dy)$. Applichiamo il **teorema di Green** (notebook 7.23) a **ciascuno** dei due integrali curvilinei reali, con $D$ la regione racchiusa da $\gamma$:

$$\oint_\gamma(u\,dx-v\,dy) = \iint_D\left(\frac{\partial(-v)}{\partial x}-\frac{\partial u}{\partial y}\right)dA = \iint_D\left(-\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}\right)dA$$

$$\oint_\gamma(v\,dx+u\,dy) = \iint_D\left(\frac{\partial u}{\partial x}-\frac{\partial v}{\partial y}\right)dA$$

Ma $f$ è olomorfa, quindi soddisfa le **equazioni di Cauchy-Riemann** (notebook 9.7): $\partial u/\partial x=\partial v/\partial y$ e $\partial u/\partial y=-\partial v/\partial x$. Sostituendo nella prima espressione: $-\partial v/\partial x-\partial u/\partial y=-\partial v/\partial x-(-\partial v/\partial x)=0$. Nella seconda: $\partial u/\partial x-\partial v/\partial y=\partial u/\partial x-\partial u/\partial x=0$. **Entrambi** gli integrali doppi sono nulli, quindi $\oint_\gamma f\,dz=0+i\times0=0$. $\blacksquare$

**Nota**: questa dimostrazione richiede che $u,v$ abbiano derivate parziali continue (per applicare Green); la versione completa del teorema (dovuta a Goursat) non richiede questa ipotesi aggiuntiva, ma la dimostrazione è più tecnica e la omettiamo. Il collegamento che questa dimostrazione rivela — tra le equazioni di Cauchy-Riemann e l'annullamento tramite Green — è comunque il cuore concettuale del risultato.

### Conseguenza: primitive ed esistenza

> **Corollario.** Se $f$ è olomorfa su un dominio semplicemente connesso $D$, allora $f$ ammette una **primitiva** $F$ (cioè $F'=f$) su $D$, e per ogni curva $\gamma$ da $z_0$ a $z_1$ in $D$: $\int_\gamma f\,dz=F(z_1)-F(z_0)$.

**Idea della dimostrazione**: fissato un punto base $a\in D$, si definisce $F(z)=\int_a^zf(w)\,dw$ (l'integrale lungo una curva qualsiasi da $a$ a $z$ — ben definito, cioè **indipendente dal percorso scelto**, esattamente perché il teorema di Cauchy garantisce che due percorsi diversi con gli stessi estremi danno lo stesso integrale, essendo la loro differenza l'integrale lungo una curva chiusa, che è zero). Si dimostra poi che $F'(z)=f(z)$, con un argomento analogo al teorema fondamentale del calcolo (notebook 5.16). Questo giustifica **rigorosamente** la scorciatoia usata nell'esercizio svolto del notebook 9.8.

### Riesame: perché $1/z$ è un'eccezione

Nel notebook 9.8, $\oint_{|z|=1}\dfrac1z\,dz=2\pi i\ne0$. Questo **non contraddice** il teorema di Cauchy: $f(z)=1/z$ è olomorfa su $\mathbb C\setminus\{0\}$, ma quel dominio **non è semplicemente connesso** rispetto a curve che circondano l'origine (l'origine è un "buco" che la curva $|z|=1$ circonda) — l'ipotesi del teorema fallisce esattamente in questo caso, e infatti l'integrale non si annulla. Questa eccezione, lungi dall'essere un problema, sarà il punto di partenza del teorema dei residui (notebook 9.14).

````python
import sympy as sp

t = sp.symbols('t', real=True)

# Verifica del teorema di Cauchy: f(z)=z^2 (olomorfa su tutto C, quindi su qualsiasi dominio) lungo un cerchio
gamma = sp.exp(sp.I * t)
gamma_prime = sp.diff(gamma, t)
f = gamma**2
integrale = sp.integrate(sp.simplify(f * gamma_prime), (t, 0, 2*sp.pi))
print("Integrale di z^2 lungo |z|=1:", integrale)

# f(z) = e^z (intera): lungo lo stesso cerchio
f2 = sp.exp(gamma)
integranda2 = sp.expand(f2 * gamma_prime, complex=True)
integrale2 = sp.integrate(integranda2, (t, 0, 2*sp.pi))
print("Integrale di e^z lungo |z|=1:", sp.simplify(integrale2))

# Verifica diretta delle equazioni di Cauchy-Riemann per f(z)=z^2, e che l'annullamento di Green segue da esse
x, y = sp.symbols('x y', real=True)
u = x**2 - y**2
v = 2*x*y
eq1 = sp.diff(u, x) - sp.diff(v, y)
eq2 = sp.diff(u, y) + sp.diff(v, x)
print("\nCauchy-Riemann per z^2:", eq1, eq2, "(entrambe nulle: coerente con l'integrale nullo)")
````

## Esercizio svolto e commentato

**Problema:** usa il teorema di Cauchy (e il suo corollario) per calcolare $\displaystyle\oint_\gamma\cos(z)\,dz$, dove $\gamma$ è **qualsiasi** curva chiusa in $\mathbb C$.

**Ragionamento:**

**Passo 1 — verifichiamo che $f(z)=\cos z$ è olomorfa su tutto $\mathbb C$.** Definendo $\cos z$ tramite la formula di Eulero inversa (notebook 9.3), $\cos z=\dfrac{e^{iz}+e^{-iz}}2$: essendo somma e composizione di funzioni intere ($e^{iz}$ è composizione dell'esponenziale intero con la funzione lineare $iz$, olomorfa per le regole trasferite dal caso reale, notebook 9.6), $\cos z$ è **intera**.

**Passo 2 — applichiamo il teorema di Cauchy** con $D=\mathbb C$ (semplicemente connesso, essendo l'intero piano, senza "buchi"): poiché $\gamma$ è una curva chiusa **qualsiasi** contenuta in $D=\mathbb C$, e $f$ è olomorfa su tutto $D$, il teorema si applica direttamente.

**Conclusione:** $\displaystyle\oint_\gamma\cos(z)\,dz=0$, **indipendentemente** dalla forma specifica di $\gamma$ (un cerchio, un quadrato, una curva qualsiasi purché chiusa).

**Perché questo esempio, apparentemente "banale", cattura la vera potenza del teorema di Cauchy:** senza il teorema, per rispondere a questa domanda avremmo dovuto **conoscere esplicitamente** la parametrizzazione di $\gamma$ e calcolare un integrale — un compito impossibile se $\gamma$ non è specificata in dettaglio ("qualsiasi curva chiusa"). Il teorema di Cauchy risponde **istantaneamente**, con una sola verifica (l'olomorfia di $f$ sull'intero dominio racchiuso da $\gamma$), a una domanda che riguarda **infinite** curve diverse contemporaneamente — è esattamente questo potere di "universalità" che rende il teorema uno strumento così centrale, piuttosto che un fatto isolato su una curva specifica.

````python
import sympy as sp

t = sp.symbols('t', real=True)

# Verifica per una curva specifica: cerchio di raggio 3 centrato in i
gamma = 3 * sp.exp(sp.I * t) + sp.I
gamma_prime = sp.diff(gamma, t)
f = sp.cos(gamma)

integranda = sp.expand(f * gamma_prime, complex=True)
integrale = sp.integrate(integranda, (t, 0, 2*sp.pi))
print("Integrale di cos(z) lungo un cerchio di raggio 3 centrato in i:", sp.simplify(integrale))
````

## Riepilogo finale

- Il **teorema di Cauchy**: se $f$ è olomorfa su un dominio semplicemente connesso, l'integrale lungo ogni curva chiusa in quel dominio è **zero** — dimostrato collegando le equazioni di Cauchy-Riemann (notebook 9.7) al teorema di Green (notebook 7.23).
- **Corollario**: funzioni olomorfe su domini semplicemente connessi ammettono sempre una primitiva; l'integrale dipende solo dagli estremi, non dal percorso.
- L'eccezione $1/z$ (integrale $2\pi i\ne0$ attorno all'origine) non contraddice il teorema: il dominio $\mathbb C\setminus\{0\}$ non è semplicemente connesso rispetto a curve che circondano l'origine.
- Il teorema risponde istantaneamente a domande su infinite curve diverse, con una sola verifica di olomorfia sul dominio racchiuso.
