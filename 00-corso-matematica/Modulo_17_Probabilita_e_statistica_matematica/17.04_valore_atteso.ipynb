# 17.4 — Valore atteso

**Prerequisiti:** 17.3 Distribuzioni, 13.4 Integrale di Lebesgue, 13.8 Spazi Lᵖ.


## Perché ci interessa

Una variabile aleatoria descrive un esito incerto, ma spesso vogliamo riassumerla con un solo numero: "in media, cosa ci aspettiamo?" Questa domanda ha già, implicitamente, una risposta pronta: l'integrale di Lebesgue del notebook 13.4 è **letteralmente** costruito per fare una media pesata di valori, pesata dalla misura degli insiemi dove quei valori vengono assunti. Il **valore atteso** non è altro che l'integrale di Lebesgue di $X$ rispetto a $P$ — un'altra istanza del principio, già visto nel notebook 17.1, che la probabilità non ha bisogno di una teoria a parte.

## Teoria, passo dopo passo

### Definizione

Se $X\ge0$, il **valore atteso** è $E[X]=\displaystyle\int_\Omega X\,dP$ (integrale di Lebesgue, notebook 13.4), sempre ben definito in $[0,+\infty]$. Per $X$ generica, scrivendo $X=X^+-X^-$ (parte positiva e negativa, una decomposizione standard nella costruzione dell'integrale di Lebesgue, notebook 13.4), si pone $E[X]=E[X^+]-E[X^-]$, definito quando **almeno uno** dei due addendi è finito; $X$ si dice **integrabile** (o si dice che $E[X]$ esiste finito) se $E[X^+]<\infty$ e $E[X^-]<\infty$, equivalentemente se $E[|X|]<\infty$, cioè $X\in L^1(\Omega,P)$ (notebook 13.8).

### Linearità

**Teorema.** $E[aX+bY]=aE[X]+bE[Y]$ per $X,Y\in L^1(\Omega,P)$ e $a,b\in\mathbb R$.

Questa è esattamente la linearità dell'integrale di Lebesgue (notebook 13.4, dimostrata per funzioni semplici nell'esercizio 2 di quel notebook, poi estesa per limiti monotoni a funzioni non negative generiche, e infine a funzioni integrabili generiche tramite $X=X^+-X^-$): nessuna dimostrazione nuova richiesta, solo un cambio di notazione ($E$ al posto di $\int\cdot\,dP$).

### Il teorema del valore atteso inconsapevole (legge dello statistico inconsapevole)

Spesso non vogliamo calcolare $E[X]$ direttamente su $\Omega$ (che può essere astratto e scomodo), ma tramite la distribuzione $\mu_X$ su $\mathbb R$ (notebook 17.2).

**Teorema.** Per $g:\mathbb R\to\mathbb R$ misurabile (notebook 13.3) tale che $g(X)$ sia integrabile,
$$E[g(X)]=\int_{\mathbb R}g(x)\,d\mu_X(x).$$

**Dimostrazione (schema, per approssimazione).** Per $g=\mathbb 1_B$ (indicatrice di un boreliano $B$): $E[\mathbb 1_B(X)]=E[\mathbb 1_{X^{-1}(B)}]=P(X^{-1}(B))=\mu_X(B)=\int_{\mathbb R}\mathbb 1_B\,d\mu_X$ (usando la definizione di $\mu_X$, notebook 17.2, e il fatto che l'integrale di un'indicatrice è la misura dell'insieme, notebook 13.4). Per $g$ semplice, $g=\sum_ic_i\mathbb 1_{B_i}$, l'uguaglianza segue per linearità (appena vista) applicata a entrambi i lati. Per $g\ge0$ generica, approssimando con funzioni semplici crescenti $g_n\uparrow g$ (la stessa tecnica di costruzione dell'integrale, notebook 13.4) e passando al limite con il teorema di convergenza monotona (notebook 13.7, applicato sia all'integrale su $\Omega$ sia a quello su $\mathbb R$): l'uguaglianza si estende a $g\ge0$, e infine a $g$ generica tramite $g=g^+-g^-$. $\blacksquare$

**Conseguenza pratica.** Per $X$ discreta con PMF $p_X$: $E[g(X)]=\sum_ig(x_i)p_X(x_i)$ (l'integrale rispetto a $\mu_X$ si riduce a una somma, essendo $\mu_X$ concentrata sui punti $x_i$). Per $X$ con densità $f_X$ (notebook 17.3): $E[g(X)]=\int_{\mathbb R}g(x)f_X(x)\,dx$ (l'integrale rispetto a $\mu_X=f_X\,d\lambda$ diventa un integrale rispetto a Lebesgue, pesato da $f_X$). In particolare, con $g(x)=x$: $E[X]=\sum_ix_ip_X(x_i)$ oppure $E[X]=\int xf_X(x)\,dx$ — le formule familiari, ora derivate (non postulate) dalla definizione astratta $E[X]=\int_\Omega X\,dP$.

### Esempi

- **Bernoulli** $\mathrm{Be}(p)$: $E[X]=1\cdot p+0\cdot(1-p)=p$.
- **Uniforme** $\mathrm U(a,b)$: $E[X]=\displaystyle\int_a^bx\cdot\frac1{b-a}\,dx=\frac1{b-a}\Big[\frac{x^2}2\Big]_a^b=\frac{a+b}2$ (il punto medio, come atteso per simmetria).
- **Esponenziale** $\mathrm{Exp}(\lambda)$: $E[X]=\displaystyle\int_0^\infty x\lambda e^{-\lambda x}\,dx=\frac1\lambda$ (integrazione per parti, notebook 5.17, che l'esercizio svolto sviluppa nel dettaglio).
- **Normale** $\mathrm N(\mu,\sigma^2)$: $E[X]=\mu$ (già verificato nel notebook 17.3, esercizio 4).

```python
import numpy as np
from scipy import integrate

# Verifica numerica di E[X] per l'esponenziale, confrontando l'integrale
# con la formula esatta 1/lambda, e verifica della linearita' E[aX+b]=aE[X]+b.

lam = 2.0

integrando = lambda x: x * lam * np.exp(-lam * x)
E_X, _ = integrate.quad(integrando, 0, np.inf)
print(f"E[X] per Exp(lambda={lam}): calcolato = {E_X:.6f}, formula 1/lambda = {1/lam:.6f}")

# Verifica di linearita' con una simulazione Monte Carlo
np.random.seed(0)
campioni = np.random.exponential(scale=1/lam, size=1000000)
a, b = 3.0, 5.0
E_X_simulata = np.mean(campioni)
E_aXb_simulata = np.mean(a*campioni + b)
E_aXb_formula = a*E_X_simulata + b

print(f"\nE[X] simulato: {E_X_simulata:.4f}")
print(f"E[aX+b] simulato: {E_aXb_simulata:.4f}")
print(f"a*E[X]+b (linearita'): {E_aXb_formula:.4f}")
```

## Esercizio svolto e commentato

**Problema:** calcola $E[X]$ per $X\sim\mathrm{Exp}(\lambda)$ usando l'integrazione per parti (notebook 5.17).

**Ragionamento:**

**Passo 1 — impostiamo l'integrale.** $E[X]=\displaystyle\int_0^\infty x\lambda e^{-\lambda x}\,dx$ (legge dello statistico inconsapevole, teoria, con densità $f_X(x)=\lambda e^{-\lambda x}$).

**Passo 2 — scegliamo le parti per l'integrazione per parti.** Poniamo $u=x$ (quindi $du=dx$) e $dv=\lambda e^{-\lambda x}dx$ (quindi $v=-e^{-\lambda x}$): la scelta è guidata dal fatto che derivare $u=x$ lo semplifica a una costante, eliminando la $x$ che rende l'integrale non elementare.

**Passo 3 — applichiamo la formula.** $\displaystyle\int_0^\infty x\lambda e^{-\lambda x}\,dx=\Big[-xe^{-\lambda x}\Big]_0^\infty+\int_0^\infty e^{-\lambda x}\,dx$.

**Passo 4 — valutiamo il termine di bordo.** $\lim_{M\to\infty}(-Me^{-\lambda M})=0$ (ogni esponenziale con base $>1$ cresce definitivamente più velocemente di qualunque potenza fissa, notebook 3.13, quindi $Me^{-\lambda M}=M/e^{\lambda M}\to0$), e in $x=0$ il termine è $0$: il termine di bordo si annulla completamente.

**Passo 5 — calcoliamo l'integrale rimanente.** $\displaystyle\int_0^\infty e^{-\lambda x}\,dx=\Big[-\frac1\lambda e^{-\lambda x}\Big]_0^\infty=0-\Big(-\frac1\lambda\Big)=\frac1\lambda$ (stesso argomento di annullamento del notebook 17.3, esercizio 1).

**Conclusione:** $E[X]=0+\dfrac1\lambda=\dfrac1\lambda$. Interpretazione: se $\lambda$ è il "tasso" di un evento (ad esempio il numero medio di eventi per unità di tempo in un processo di Poisson, che il notebook 17.10 approfondirà), il tempo medio di attesa fino al prossimo evento è $1/\lambda$ — coerente con l'intuizione (tasso alto, attesa breve).

```python
import sympy as sp

x, lam = sp.symbols('x lambda', positive=True)
f_X = lam * sp.exp(-lam * x)

E_X = sp.integrate(x * f_X, (x, 0, sp.oo))
print("E[X] per Exp(lambda):", sp.simplify(E_X))

# Verifica con integrazione per parti esplicita
u, dv = x, lam*sp.exp(-lam*x)
v = sp.integrate(dv, x)
termine_bordo = sp.limit(-x*sp.exp(-lam*x), x, sp.oo) - (-0*sp.exp(0))
integrale_residuo = sp.integrate(sp.exp(-lam*x), (x, 0, sp.oo))
print("Termine di bordo:", termine_bordo)
print("Integrale residuo:", integrale_residuo)
```

4. *(più impegnativo)* Dimostra la **disuguaglianza di Jensen** per $g$ **convessa** (una nozione nuova: $g''\ge0$ ovunque, che geometricamente significa che il grafico di $g$ sta sempre sopra ogni sua retta tangente): $g(E[X])\le E[g(X)]$. Suggerimento: usa che una funzione convessa sta sempre sopra ogni sua retta tangente, $g(x)\ge g(E[X])+g'(E[X])(x-E[X])$ per ogni $x$ (una conseguenza della formula di Taylor con resto, notebook 5.14, applicata al punto $E[X]$), poi applica $E[\cdot]$ a entrambi i lati usando la monotonia e la linearità dell'integrale.

**1.** $E[X]=\sum_{k=0}^\infty k\dfrac{\lambda^k}{k!}e^{-\lambda}=\sum_{k=1}^\infty\dfrac{\lambda^k}{(k-1)!}e^{-\lambda}$ (il termine $k=0$ è nullo, e per $k\ge1$, $k/k!=1/(k-1)!$). Ponendo $j=k-1$: $=\lambda\sum_{j=0}^\infty\dfrac{\lambda^j}{j!}e^{-\lambda}=\lambda e^\lambda e^{-\lambda}=\lambda$ (riconoscendo $\sum_j\lambda^j/j!=e^\lambda$, la serie esponenziale, notebook 5.20). Quindi $E[X]=\lambda$: il parametro $\lambda$ della Poisson è, coerentemente col suo ruolo di "tasso medio", esattamente il suo valore atteso.

**2.** $E[X^2]=\displaystyle\int_0^1x^2\cdot1\,dx=\Big[\frac{x^3}3\Big]_0^1=\frac13$.

**3.** $\displaystyle\int_0^\infty\frac{x}{\pi(1+x^2)}\,dx$: per $x$ grande, $\dfrac{x}{1+x^2}\sim\dfrac1x$ (dividendo numeratore e denominatore per $x^2$: $\frac{1/x}{1/x^2+1}\to\frac1x$ per confronto asintotico), e $\int^\infty\frac1x\,dx=+\infty$ (un fatto standard sugli integrali impropri, notebook 5.18: la funzione $1/x$ non è integrabile all'infinito). Per un criterio del confronto asintotico tra integrali impropri (variante della stessa idea usata per le serie, notebook 5.19, applicata qui agli integrali), anche $\int_0^\infty\frac{x}{\pi(1+x^2)}\,dx=+\infty$. Quindi $E[X^+]=+\infty$, e per simmetria (la densità di Cauchy è pari, notebook 3.8) anche $E[X^-]=+\infty$: essendo entrambe le parti infinite, $E[X]$ **non è definito** — un classico controesempio di variabile aleatoria "patologica" senza valore atteso, nonostante abbia una densità perfettamente legittima.

**4.** Per $g$ convessa, la disuguaglianza $g(x)\ge g(E[X])+g'(E[X])(x-E[X])$ per ogni $x$ segue dalla formula di Taylor con resto di Lagrange (notebook 5.14) attorno al punto $x_0=E[X]$: $g(x)=g(x_0)+g'(x_0)(x-x_0)+\frac12g''(\xi)(x-x_0)^2$ per qualche $\xi$ tra $x_0$ e $x$, e l'ultimo termine è $\ge0$ essendo $g''\ge0$ per convessità. Sostituendo $X$ al posto di $x$ (una variabile aleatoria, non un numero fissato) e applicando $E[\cdot]$ a entrambi i lati della disuguaglianza $g(X)\ge g(E[X])+g'(E[X])(X-E[X])$ (monotonia dell'integrale di Lebesgue rispetto all'ordinamento delle funzioni, un fatto immediato dalla costruzione dell'integrale, notebook 13.4): $E[g(X)]\ge g(E[X])+g'(E[X])\cdot E[X-E[X]]$. Ma $E[X-E[X]]=E[X]-E[X]=0$ (linearità, teoria), quindi il secondo termine si annulla, lasciando $E[g(X)]\ge g(E[X])$: la disuguaglianza di Jensen.

## Riepilogo finale

- Il **valore atteso** $E[X]=\int_\Omega X\,dP$ è semplicemente l'integrale di Lebesgue di $X$ (notebook 13.4): nessuna teoria nuova, solo un nome e un'interpretazione probabilistica.
- $X$ ha valore atteso finito se e solo se $X\in L^1(\Omega,P)$ (notebook 13.8).
- La **legge dello statistico inconsapevole** permette di calcolare $E[g(X)]$ direttamente dalla distribuzione di $X$ su $\mathbb R$, senza tornare a $\Omega$: nel caso discreto una somma, nel caso continuo un integrale rispetto alla densità.
- La linearità di $E$ è ereditata, senza nuova dimostrazione, dalla linearità dell'integrale di Lebesgue.
- La disuguaglianza di Jensen, conseguenza della formula di Taylor applicata alla convessità, è uno strumento generale per confrontare $E[g(X)]$ con $g(E[X])$ ogni volta che $g$ è convessa.
