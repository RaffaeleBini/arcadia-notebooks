# 17.7 — Convergenza di variabili aleatorie

**Prerequisiti:** 17.5 Varianza, 13.5 Convergenza quasi ovunque, 13.6 Convergenza in misura, 13.8 Spazi Lᵖ.


## Perché ci interessa

I prossimi due notebook (legge dei grandi numeri, teorema centrale del limite) affermano che, sotto ipotesi opportune, una successione di variabili aleatorie "si avvicina" a un limite. Ma il notebook 13.5 ha già mostrato che esistono **più nozioni distinte** di "avvicinarsi" per successioni di funzioni, che in generale non sono equivalenti. Questo notebook riprende esattamente quelle nozioni — quasi ovunque, in misura, in $L^p$ (notebook 13.5, 13.6, 13.8) — le rilegge nel linguaggio probabilistico, e ne aggiunge una nuova, specifica della probabilità: la **convergenza in distribuzione**, necessaria perché il teorema centrale del limite parla del comportamento della *distribuzione*, non dei valori puntuali, di una successione di variabili aleatorie.

## Teoria, passo dopo passo

### Convergenza quasi certa

$X_n\to X$ **quasi certamente** (o **quasi sicuramente**, abbreviato q.c.) se $X_n\to X$ **quasi ovunque** rispetto a $P$ (notebook 13.5): $P\big(\{\omega:X_n(\omega)\to X(\omega)\}\big)=1$. Nessuna definizione nuova: lo stesso concetto del Modulo 13, con $\mu=P$.

### Convergenza in probabilità

$X_n\to X$ **in probabilità** se $X_n\to X$ **in misura** rispetto a $P$ (notebook 13.6): per ogni $\varepsilon>0$, $P(|X_n-X|\ge\varepsilon)\to0$.

**Teorema (ereditato dal Modulo 13).** La convergenza quasi certa implica la convergenza in probabilità.

Questo è esattamente il teorema del notebook 13.6 ("convergenza q.o. su misura finita implica convergenza in misura"), applicato a $\mu=P$: essendo $P(\Omega)=1<\infty$ **sempre**, l'ipotesi di misura finita è automaticamente soddisfatta in ogni spazio di probabilità — nessuna verifica aggiuntiva richiesta. L'implicazione inversa è falsa in generale: lo stesso controesempio "a macchina da scrivere" del notebook 13.6, reinterpretato con $P=$ misura di Lebesgue su $[0,1]$, fornisce una successione che converge in probabilità ma non converge quasi certamente in nessun punto.

### Convergenza in $L^p$

$X_n\to X$ **in $L^p$** ($p\ge1$) se $E[|X_n-X|^p]\to0$, cioè $X_n\to X$ nella norma di $L^p(\Omega,P)$ (notebook 13.8). I due casi più usati sono $p=1$ (convergenza in media) e $p=2$ (convergenza in media quadratica).

**Teorema.** La convergenza in $L^p$ (per qualunque $p\ge1$) implica la convergenza in probabilità.

**Dimostrazione.** Per la disuguaglianza di Markov (notebook 17.5, esercizio 4, applicata a $Z=|X_n-X|^p\ge0$ e $t=\varepsilon^p$): $P(|X_n-X|^p\ge\varepsilon^p)\le\dfrac{E[|X_n-X|^p]}{\varepsilon^p}$. Ma $|X_n-X|^p\ge\varepsilon^p\iff|X_n-X|\ge\varepsilon$ (radice $p$-esima, funzione crescente su $[0,\infty)$, notebook 3.9), quindi $P(|X_n-X|\ge\varepsilon)\le\dfrac{E[|X_n-X|^p]}{\varepsilon^p}\to0$ per ipotesi. $\blacksquare$

**Nessuna relazione generale tra $L^p$ e q.c.** Convergenza in $L^p$ non implica convergenza q.c. (lo stesso controesempio "a macchina da scrivere" del notebook 13.6 converge anche in $L^p$ per ogni $p$, essendo le funzioni limitate su un dominio di misura decrescente a zero, ma non converge q.c. in nessun punto), e nemmeno viceversa in generale (esistono successioni convergenti q.c. ma non in $L^p$, se $|X_n|^p$ non è uniformemente controllata: un fenomeno legato all'ipotesi del teorema di convergenza dominata, notebook 13.7, che garantisce convergenza in $L^1$ **solo** quando è disponibile una maggiorante integrabile).

### Convergenza in distribuzione

Questa è una nozione **nuova**, specifica della probabilità, più debole di tutte le precedenti: $X_n\to X$ **in distribuzione** (scritto $X_n\xrightarrow{d}X$) se $F_{X_n}(x)\to F_X(x)$ per ogni $x$ in cui $F_X$ è **continua** (notebook 17.2). La restrizione ai punti di continuità è essenziale: nei punti di salto di $F_X$, piccoli spostamenti della distribuzione limite possono far oscillare $F_{X_n}(x)$ senza che questo rifletta un vero "mancato avvicinamento" (l'esercizio 4 costruisce un esempio esplicito).

**Perché è più debole.** La convergenza in distribuzione riguarda solo le distribuzioni $\mu_{X_n},\mu_X$, non richiede che $X_n$ e $X$ siano nemmeno definite sullo stesso spazio di probabilità $\Omega$ — un salto concettuale importante rispetto alle tre nozioni precedenti, che confrontano $X_n(\omega)$ e $X(\omega)$ punto per punto (o quasi punto per punto) sullo **stesso** $\Omega$.

**Teorema.** La convergenza in probabilità implica la convergenza in distribuzione.

**Dimostrazione (schema).** Per $x$ punto di continuità di $F_X$ e $\varepsilon>0$: $\{X_n\le x\}\subseteq\{X\le x+\varepsilon\}\cup\{|X_n-X|>\varepsilon\}$ (se $X_n(\omega)\le x$ ma $X(\omega)>x+\varepsilon$, allora necessariamente $|X_n(\omega)-X(\omega)|>\varepsilon$). Per subadditività e monotonia (notebook 17.1): $F_{X_n}(x)\le F_X(x+\varepsilon)+P(|X_n-X|>\varepsilon)$. Passando al $\limsup$ per $n\to\infty$ (il secondo termine tende a $0$ per convergenza in probabilità): $\limsup_nF_{X_n}(x)\le F_X(x+\varepsilon)$. Un argomento simmetrico dà $\liminf_nF_{X_n}(x)\ge F_X(x-\varepsilon)$. Facendo $\varepsilon\to0$ e usando la continuità di $F_X$ in $x$: entrambi i limiti tendono a $F_X(x)$, quindi $F_{X_n}(x)\to F_X(x)$. $\blacksquare$

### Riepilogo delle implicazioni

$$\text{q.c.}\ \Longrightarrow\ \text{in probabilità}\ \Longleftarrow\ \text{in }L^p\qquad\text{in probabilità}\ \Longrightarrow\ \text{in distribuzione}$$

Nessuna di queste frecce si inverte in generale: quattro nozioni genuinamente diverse, con la convergenza in distribuzione la più debole di tutte — ed è proprio per questo che il teorema centrale del limite (notebook 17.9), che afferma solo convergenza in distribuzione, è un risultato più delicato di quanto sembri: non dice nulla sul comportamento puntuale delle variabili coinvolte.

```python
import numpy as np
import matplotlib.pyplot as plt

# Ricostruiamo il controesempio "a macchina da scrivere" (gia' visto nel
# notebook 13.6) nel linguaggio della probabilita': su Omega=[0,1] con P=Lebesgue,
# X_n = indicatrice di un blocco che si restringe, converge in probabilita' a 0
# ma NON converge quasi certamente in nessun punto.

def blocchi_macchina_da_scrivere(n_max):
    blocchi = []
    k = 0
    while len(blocchi) < n_max:
        n_blocchi_ciclo = 2**k
        larghezza = 1 / n_blocchi_ciclo
        for i in range(n_blocchi_ciclo):
            if len(blocchi) >= n_max:
                break
            blocchi.append((i * larghezza, (i + 1) * larghezza))
        k += 1
    return blocchi

blocchi = blocchi_macchina_da_scrivere(30)

# Verifica: P(|X_n - 0| >= eps) = larghezza del blocco n-esimo -> 0 (convergenza in probabilita')
larghezze = [b - a for a, b in blocchi]
print("Larghezze dei primi 10 blocchi (= P(X_n=1), tendono a 0):")
print([f"{w:.4f}" for w in larghezze[:10]])

# Verifica: per un punto fissato omega, X_n(omega) vale 1 infinite volte (non converge)
omega_fissato = 0.37
valori_X_n = [1 if a <= omega_fissato < b else 0 for a, b in blocchi]
print(f"\nValori di X_n({omega_fissato}) per n=1..30:")
print(valori_X_n)
print("Il valore 1 ricompare infinite volte: nessuna convergenza puntuale in questo punto.")
```

## Esercizio svolto e commentato

**Problema:** sia $X_n\sim\mathrm N(0,1/n)$ per $n=1,2,\ldots$ Dimostra che $X_n\to0$ sia in $L^2$ sia in distribuzione, identificando il limite in distribuzione.

**Ragionamento:**

**Passo 1 — calcoliamo $E[X_n^2]$.** Per una normale $\mathrm N(0,\sigma^2)$, $E[X^2]=\mathrm{Var}(X)+E[X]^2=\sigma^2+0=\sigma^2$ (formula del notebook 17.5, con $E[X]=0$ per simmetria). Qui $\sigma^2=1/n$, quindi $E[X_n^2]=1/n$.

**Passo 2 — verifichiamo la convergenza in $L^2$.** $E[|X_n-0|^2]=E[X_n^2]=1/n\to0$ per $n\to\infty$ (notebook 5.4, limite di una successione elementare): $X_n\to0$ in $L^2$.

**Passo 3 — deduciamo la convergenza in probabilità (teoria).** La convergenza in $L^2$ implica la convergenza in probabilità: $X_n\to0$ in probabilità.

**Passo 4 — deduciamo la convergenza in distribuzione (teoria).** La convergenza in probabilità implica la convergenza in distribuzione: $X_n\xrightarrow{d}0$, intendendo con questo che $F_{X_n}(x)\to F_{\delta_0}(x)$ per ogni $x\ne0$ (punto di continuità della funzione di ripartizione della variabile costante $0$, che vale $0$ per $x<0$ e $1$ per $x\ge0$ — un salto proprio in $x=0$, l'unico punto escluso dal teorema).

**Conclusione:** man mano che $n$ cresce, la normale $\mathrm N(0,1/n)$ si concentra sempre più attorno a $0$ (la sua varianza tende a zero), e questo si traduce, in ogni linguaggio di convergenza disponibile (tranne quello quasi certo, che qui non ha nemmeno senso discutere: le $X_n$ potrebbero essere definite su spazi di probabilità diversi, non essendo richiesta alcuna relazione tra loro oltre alla legge), nella convergenza verso la variabile aleatoria costante $0$.

```python
import numpy as np

np.random.seed(4)
n_valori = [1, 5, 20, 100, 1000]

print("Verifica di E[X_n^2] -> 0 (convergenza L^2) per X_n ~ N(0, 1/n):")
for n in n_valori:
    campioni = np.random.normal(0, np.sqrt(1/n), size=500000)
    E_X2 = np.mean(campioni**2)
    print(f"  n={n}: E[X_n^2] simulato = {E_X2:.5f}  (teorico: {1/n:.5f})")

print("\nVerifica di convergenza in distribuzione: F_{X_n}(0.1) -> 1 (massa si concentra vicino a 0)")
for n in n_valori:
    campioni = np.random.normal(0, np.sqrt(1/n), size=500000)
    F_n_01 = np.mean(campioni <= 0.1)
    print(f"  n={n}: P(X_n <= 0.1) = {F_n_01:.4f}")
```

Suggerimento: per $\varepsilon>0$, $P(|X_n-c|\ge\varepsilon)=P(X_n\le c-\varepsilon)+P(X_n\ge c+\varepsilon)\le F_{X_n}(c-\varepsilon)+\big(1-F_{X_n}(c+\varepsilon/2)\big)$, e usa che $c-\varepsilon,c+\varepsilon/2$ sono punti di continuità della funzione di ripartizione della costante $c$ (che ha un unico salto, in $c$ stesso).

## Riepilogo finale

- La convergenza **quasi certa** e **in probabilità** sono, rispettivamente, la convergenza quasi ovunque e in misura del Modulo 13, applicate a $P$; in uno spazio di probabilità (misura finita per definizione) la prima implica sempre la seconda.
- La convergenza **in $L^p$** implica quella in probabilità, via la disuguaglianza di Markov (notebook 17.5).
- La convergenza **in distribuzione**, la più debole delle quattro, riguarda solo le distribuzioni $\mu_{X_n}$, non richiede nemmeno che le $X_n$ vivano sullo stesso spazio di probabilità, ed è implicata dalla convergenza in probabilità.
- Un'eccezione notevole: se il limite è una **costante**, la convergenza in distribuzione implica anche quella in probabilità — un fatto che useremo esplicitamente nella legge dei grandi numeri.
- Nessuna di queste implicazioni si inverte in generale: quattro nozioni di "avvicinarsi" genuinamente distinte, ciascuna adatta a contesti diversi.
