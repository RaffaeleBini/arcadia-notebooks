# 17.6 — Indipendenza

**Prerequisiti:** 17.1 Probabilità assiomatica, 17.2 Variabili aleatorie, 13.9 Misure prodotto.


## Perché ci interessa

Finora abbiamo trattato $X$ e $Y$ come oggetti isolati. Ma la probabilità diventa davvero utile quando descrive **più** variabili insieme: due lanci di moneta, il rumore su due sensori diversi, due misure ripetute. La nozione chiave che rende trattabili questi casi è l'**indipendenza**: informalmente, "sapere il valore di $X$ non cambia nulla sulla probabilità di $Y$".

Il notebook 13.9 ha costruito la **misura prodotto** e dimostrato il teorema di Fubini-Tonelli proprio per poter integrare su spazi che sono un prodotto di due fattori. L'indipendenza probabilistica è esattamente il caso in cui la distribuzione congiunta di $(X,Y)$ **è** una misura prodotto: gli stessi strumenti del Modulo 13, applicati qui per la prima volta al di fuori della pura teoria della misura.

## Teoria, passo dopo passo

### Eventi indipendenti

Due eventi $A,B\in\mathcal F$ sono **indipendenti** se $P(A\cap B)=P(A)P(B)$. Una famiglia di eventi $A_1,\ldots,A_n$ è (mutuamente) indipendente se, per ogni sottofamiglia $A_{i_1},\ldots,A_{i_k}$, $P(A_{i_1}\cap\cdots\cap A_{i_k})=P(A_{i_1})\cdots P(A_{i_k})$ — una condizione più forte della sola indipendenza a coppie (esiste un controesempio classico, sviluppato nell'esercizio 3, con eventi indipendenti a due a due ma non mutuamente indipendenti).

### Variabili aleatorie indipendenti

$X$ e $Y$ sono **indipendenti** se, per ogni coppia di boreliani $A,B\in\mathcal B(\mathbb R)$, gli eventi $\{X\in A\}$ e $\{Y\in B\}$ sono indipendenti: $P(X\in A,Y\in B)=P(X\in A)P(Y\in B)$.

**Riformulazione in termini di misure prodotto.** Consideriamo la variabile aleatoria vettoriale $(X,Y):\Omega\to\mathbb R^2$, con distribuzione congiunta $\mu_{(X,Y)}(C)=P\big((X,Y)\in C\big)$ per $C\in\mathcal B(\mathbb R^2)=\mathcal B(\mathbb R)\otimes\mathcal B(\mathbb R)$ (la σ-algebra prodotto, notebook 13.9, che coincide con i boreliani di $\mathbb R^2$ generati dai rettangoli). Applicata a un rettangolo $C=A\times B$: $\mu_{(X,Y)}(A\times B)=P(X\in A,Y\in B)$.

**Teorema.** $X,Y$ sono indipendenti se e solo se $\mu_{(X,Y)}=\mu_X\otimes\mu_Y$ (la misura prodotto delle distribuzioni marginali, notebook 13.9).

**Dimostrazione.** Se $X,Y$ sono indipendenti, $\mu_{(X,Y)}(A\times B)=P(X\in A,Y\in B)=P(X\in A)P(Y\in B)=\mu_X(A)\mu_Y(B)=(\mu_X\otimes\mu_Y)(A\times B)$ per ogni rettangolo $A\times B$: le due misure coincidono sui rettangoli, che generano $\mathcal B(\mathbb R^2)$; per un fatto generale di teoria della misura (due misure di probabilità che coincidono su una famiglia chiusa per intersezioni finite e che genera la σ-algebra coincidono ovunque — un raffinamento del criterio di unicità che estende quanto visto nel notebook 13.2 per la misura di Lebesgue), $\mu_{(X,Y)}=\mu_X\otimes\mu_Y$ su tutta $\mathcal B(\mathbb R^2)$. Viceversa, se $\mu_{(X,Y)}=\mu_X\otimes\mu_Y$, allora per definizione di misura prodotto (notebook 13.9) $\mu_{(X,Y)}(A\times B)=\mu_X(A)\mu_Y(B)$ per ogni $A,B$, cioè $P(X\in A,Y\in B)=P(X\in A)P(Y\in B)$: indipendenza. $\blacksquare$

### La conseguenza più importante: $E[XY]=E[X]E[Y]$

**Teorema.** Se $X,Y$ sono indipendenti e $X,Y\in L^1$, allora $XY\in L^1$ e $E[XY]=E[X]E[Y]$.

**Dimostrazione.** Per la legge dello statistico inconsapevole (notebook 17.4) applicata alla funzione $g(x,y)=xy$ e alla distribuzione congiunta: $E[XY]=\displaystyle\int_{\mathbb R^2}xy\,d\mu_{(X,Y)}(x,y)=\int_{\mathbb R^2}xy\,d(\mu_X\otimes\mu_Y)(x,y)$ (per il teorema appena dimostrato). Per il teorema di Fubini-Tonelli (notebook 13.9), questo integrale doppio si spezza in integrali iterati: $=\displaystyle\int_{\mathbb R}\Big(\int_{\mathbb R}xy\,d\mu_Y(y)\Big)d\mu_X(x)=\int_{\mathbb R}x\Big(y\text{-integrale}\Big)d\mu_X(x)=\int_{\mathbb R}x\,E[Y]\,d\mu_X(x)=E[Y]\int_{\mathbb R}x\,d\mu_X(x)=E[Y]E[X]$ (portando fuori la costante $E[Y]$ dall'integrale interno, poi dall'integrale esterno). $\blacksquare$

**Conseguenza per la covarianza e la varianza.** Se $X,Y$ sono indipendenti: $\mathrm{Cov}(X,Y)=E[XY]-E[X]E[Y]=0$ (per il teorema appena dimostrato). Quindi, per la formula del notebook 17.5, $\mathrm{Var}(X+Y)=\mathrm{Var}(X)+\mathrm{Var}(Y)+2\mathrm{Cov}(X,Y)=\mathrm{Var}(X)+\mathrm{Var}(Y)$: **le varianze si sommano**, senza termine incrociato — il fatto che renderà possibile, nel notebook 17.8, il controllo della varianza di una somma di molte variabili indipendenti.

**Attenzione (l'implicazione inversa è falsa).** $\mathrm{Cov}(X,Y)=0$ (variabili **non correlate**) **non implica** indipendenza in generale: l'esercizio 4 costruisce un controesempio esplicito. L'indipendenza è una condizione molto più forte, che riguarda l'intera distribuzione congiunta, non solo un singolo numero riassuntivo.

```python
import numpy as np

# Verifica: per X,Y indipendenti, E[XY]=E[X]E[Y] e Cov(X,Y)=0;
# costruiamo anche un esempio di variabili NON correlate ma NON indipendenti.

np.random.seed(2)
n = 2_000_000

# Caso 1: X, Y indipendenti (generate separatamente)
X = np.random.exponential(scale=2.0, size=n)
Y = np.random.normal(loc=5.0, scale=1.0, size=n)

E_XY = np.mean(X * Y)
E_X_E_Y = np.mean(X) * np.mean(Y)
print("Caso INDIPENDENTE:")
print(f"  E[XY] = {E_XY:.4f}")
print(f"  E[X]*E[Y] = {E_X_E_Y:.4f}")
print(f"  Cov(X,Y) = {np.mean((X-np.mean(X))*(Y-np.mean(Y))):.4f}  (atteso: ~0)")

# Caso 2: Z non correlata con W ma NON indipendente
# W ~ Uniforme(-1,1), Z = W^2 : Cov(W,Z)=E[W^3]-E[W]E[W^2]=0-0=0 (simmetria)
# ma Z e' funzione deterministica di W: massima dipendenza possibile!
W = np.random.uniform(-1, 1, size=n)
Z = W**2
Cov_WZ = np.mean((W - np.mean(W)) * (Z - np.mean(Z)))
print("\nCaso NON CORRELATO ma DIPENDENTE (Z = W^2):")
print(f"  Cov(W,Z) = {Cov_WZ:.6f}  (vicino a 0, ma Z e' determinato da W!)")
print(f"  P(Z > 0.9 | W > 0.9) = {np.mean(Z[W>0.9]>0.9):.4f}  (dovrebbe essere alta, non ~P(Z>0.9))")
print(f"  P(Z > 0.9) marginale = {np.mean(Z>0.9):.4f}")
```

## Esercizio svolto e commentato

**Problema:** siano $X,Y$ indipendenti con $X\sim\mathrm U(0,1)$ e $Y\sim\mathrm U(0,1)$. Calcola $P(X+Y\le1)$ usando il teorema di Fubini-Tonelli (notebook 13.9) applicato alla misura prodotto delle due uniformi.

**Ragionamento:**

**Passo 1 — scriviamo la probabilità come un integrale sulla misura prodotto.** Per indipendenza (teoria), $\mu_{(X,Y)}=\mu_X\otimes\mu_Y$, dove $\mu_X,\mu_Y$ sono entrambe la misura di Lebesgue ristretta a $[0,1]$ (notebook 17.1). Quindi $P(X+Y\le1)=\mu_{(X,Y)}\big(\{(x,y):x+y\le1\}\big)=(\lambda\otimes\lambda)\big(\{(x,y)\in[0,1]^2:x+y\le1\}\big)$.

**Passo 2 — riconosciamo che questo è l'area di una regione del piano.** Per la misura prodotto di Lebesgue con sé stessa, $(\lambda\otimes\lambda)(C)$ per $C\subseteq[0,1]^2$ è esattamente l'**area** (bidimensionale) di $C$ nel senso ordinario (notebook 7.14, integrali multipli, di cui la misura prodotto è la formalizzazione rigorosa, notebook 13.9).

**Passo 3 — calcoliamo l'area con Fubini-Tonelli.** $\{(x,y)\in[0,1]^2:x+y\le1\}$ è il triangolo sotto la retta $y=1-x$. Per Tonelli (la funzione indicatrice è non negativa, quindi l'ordine di integrazione non conta): $\displaystyle\int_0^1\Big(\int_0^{1-x}dy\Big)dx=\int_0^1(1-x)\,dx=\Big[x-\frac{x^2}2\Big]_0^1=1-\frac12=\frac12$.

**Conclusione:** $P(X+Y\le1)=\dfrac12$ — coerente con l'intuizione geometrica elementare (il triangolo occupa esattamente metà del quadrato unitario), ma qui ottenuto rigorosamente come istanza del teorema di Fubini-Tonelli applicato alla misura prodotto di due distribuzioni indipendenti, non come un argomento puramente geometrico a parte.

```python
import numpy as np
from scipy import integrate

# Verifica numerica (simulazione) e simbolica (integrale doppio)
np.random.seed(3)
n = 2_000_000
X = np.random.uniform(0, 1, n)
Y = np.random.uniform(0, 1, n)
P_simulata = np.mean(X + Y <= 1)
print(f"P(X+Y<=1) simulata: {P_simulata:.4f}  (attesa: 0.5)")

# Integrale doppio via Fubini-Tonelli
risultato, _ = integrate.dblquad(lambda y, x: 1, 0, 1, 0, lambda x: 1-x)
print(f"Area del triangolo (integrale doppio): {risultato:.4f}")
```

## Riepilogo finale

- Due variabili aleatorie sono **indipendenti** se, e solo se, la loro distribuzione congiunta è la misura prodotto delle distribuzioni marginali (notebook 13.9): $\mu_{(X,Y)}=\mu_X\otimes\mu_Y$.
- Per variabili indipendenti, $E[XY]=E[X]E[Y]$ (via Fubini-Tonelli, notebook 13.9), quindi $\mathrm{Cov}(X,Y)=0$ e le varianze si sommano: $\mathrm{Var}(X+Y)=\mathrm{Var}(X)+\mathrm{Var}(Y)$.
- L'implicazione inversa è falsa: variabili non correlate possono essere fortemente dipendenti (come $Y=X^2$), perché la covarianza cattura solo dipendenza lineare.
- L'indipendenza mutua di più eventi è più forte della semplice indipendenza a coppie, come mostra il classico controesempio con quattro punti equiprobabili.
- L'additività delle varianze per somme di indipendenti sarà lo strumento chiave, nel notebook 17.8, per dimostrare la legge dei grandi numeri.
