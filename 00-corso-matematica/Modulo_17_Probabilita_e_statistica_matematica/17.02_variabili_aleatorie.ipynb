## Perché ci interessa

Nel notebook 17.1 abbiamo lavorato direttamente con eventi astratti $A\subseteq\Omega$. Ma quasi sempre, in pratica, non ci interessa l'esito grezzo dell'esperimento (quale faccia del dado, quale punto dell'intervallo) — ci interessa un **numero** che dipende da quell'esito: la somma di due dadi, la posizione di una particella, il guadagno di una scommessa. Serve un modo di passare da $\Omega$ (spesso astratto o scomodo) a $\mathbb R$ (dove sappiamo fare analisi).

Questo passaggio è esattamente una **funzione misurabile** $X:\Omega\to\mathbb R$ nel senso del notebook 13.3: qui la chiamiamo **variabile aleatoria**. La misurabilità non è un dettaglio tecnico da dimenticare: è precisamente la condizione che garantisce che domande come "qual è la probabilità che $X$ sia minore di $5$?" abbiano senso, cioè che $\{X<5\}$ sia effettivamente un evento a cui $P$ assegna un valore.

## Teoria, passo dopo passo

### Definizione

Una **variabile aleatoria** su uno spazio di probabilità $(\Omega,\mathcal F,P)$ è una funzione misurabile $X:\Omega\to\mathbb R$ (notebook 13.3): per ogni boreliano $B\in\mathcal B(\mathbb R)$ (notebook 13.1), $X^{-1}(B)=\{\omega\in\Omega:X(\omega)\in B\}\in\mathcal F$. Questa condizione garantisce che eventi come $\{X\le x\}=X^{-1}((-\infty,x])$, $\{X\in[a,b]\}=X^{-1}([a,b])$ siano sempre eventi legittimi, ai quali $P$ può essere applicata. Per la caratterizzazione della misurabilità tramite una famiglia generatrice (notebook 13.3), basta verificare $\{X\le x\}\in\mathcal F$ per ogni $x\in\mathbb R$: gli intervalli $(-\infty,x]$ generano $\mathcal B(\mathbb R)$ (notebook 13.1), quindi questa condizione, apparentemente più debole, è in realtà equivalente alla misurabilità piena.

### La misura indotta (distribuzione)

Data una variabile aleatoria $X$, definiamo, per ogni boreliano $B\in\mathcal B(\mathbb R)$:
$$\mu_X(B)=P\big(X^{-1}(B)\big)=P(X\in B).$$

**Fatto.** $\mu_X$ è una misura di probabilità su $(\mathbb R,\mathcal B(\mathbb R))$, chiamata **distribuzione** (o legge) di $X$.

**Dimostrazione.** $\mu_X(\emptyset)=P(X^{-1}(\emptyset))=P(\emptyset)=0$. Per la σ-additività: se $B_1,B_2,\ldots$ sono boreliani disgiunti, $X^{-1}(B_1),X^{-1}(B_2),\ldots$ sono anch'essi disgiunti (se $\omega\in X^{-1}(B_i)\cap X^{-1}(B_j)$ con $i\ne j$, allora $X(\omega)\in B_i\cap B_j=\emptyset$, assurdo — qui usiamo solo il fatto elementare, verificabile direttamente dalla definizione di preimmagine, che $X^{-1}(A\cap B)=X^{-1}(A)\cap X^{-1}(B)$), e $X^{-1}\big(\bigcup_nB_n\big)=\bigcup_nX^{-1}(B_n)$ (stessa verifica diretta, applicata a un'unione al posto di un'intersezione). Quindi $\mu_X\big(\bigcup_nB_n\big)=P\Big(\bigcup_nX^{-1}(B_n)\Big)=\sum_nP(X^{-1}(B_n))=\sum_n\mu_X(B_n)$ (σ-additività di $P$, notebook 13.2). Infine $\mu_X(\mathbb R)=P(X^{-1}(\mathbb R))=P(\Omega)=1$. $\blacksquare$

**Perché questo è un guadagno concettuale.** $\mu_X$ vive su $(\mathbb R,\mathcal B(\mathbb R))$, uno spazio fisso e concreto, indipendentemente da quanto "strano" o astratto sia lo spazio campionario originale $\Omega$. D'ora in avanti, per moltissime domande su $X$ (probabilità che cada in un certo insieme, valore atteso, varianza), potremo lavorare direttamente con $\mu_X$ su $\mathbb R$, dimenticando $\Omega$: la variabile aleatoria ha "trasportato" tutta l'informazione probabilistica rilevante da $\Omega$ a $\mathbb R$.

### La funzione di ripartizione

La **funzione di ripartizione** (o funzione di distribuzione cumulativa) di $X$ è $F_X(x)=P(X\le x)=\mu_X((-\infty,x])$.

**Teorema (proprietà di $F_X$).**
1. $F_X$ è **non decrescente**.
2. $F_X$ è **continua da destra**.
3. $\lim_{x\to-\infty}F_X(x)=0$ e $\lim_{x\to+\infty}F_X(x)=1$.

**Dimostrazione.**

**(1)** Se $x\le y$, $(-\infty,x]\subseteq(-\infty,y]$, quindi $\{X\le x\}\subseteq\{X\le y\}$, e per monotonia di $P$ (notebook 13.2, ereditata da 17.1) $F_X(x)=P(X\le x)\le P(X\le y)=F_X(y)$.

**(2)** Fissato $x$, prendiamo $x_n\downarrow x$ (una successione decrescente con limite $x$, ad esempio $x_n=x+1/n$). Gli eventi $\{X\le x_n\}$ formano una successione **decrescente** (essendo $x_n$ decrescente) con $\bigcap_n\{X\le x_n\}=\{X\le x\}$ (un punto soddisfa $X(\omega)\le x_n$ per ogni $n$ se e solo se $X(\omega)\le x$, per definizione di limite decrescente). Per continuità dall'alto (notebook 17.1, sempre valida in uno spazio di probabilità): $F_X(x_n)=P(X\le x_n)\to P(X\le x)=F_X(x)$: questo è esattamente la continuità da destra.

**(3)** Prendiamo $x_n\to-\infty$ crescente in valore assoluto negativo (cioè $x_n\downarrow-\infty$): $\{X\le x_n\}$ decresce a $\bigcap_n\{X\le x_n\}=\emptyset$ (nessun valore reale è $\le-\infty$), quindi per continuità dall'alto $F_X(x_n)\to P(\emptyset)=0$. Analogamente, per $x_n\uparrow+\infty$, $\{X\le x_n\}$ **cresce** a $\bigcup_n\{X\le x_n\}=\Omega$ (ogni valore reale di $X$ è eventualmente $\le x_n$), quindi per continuità dal basso (notebook 13.2) $F_X(x_n)\to P(\Omega)=1$. $\blacksquare$

**Fatto notevole (senza dimostrazione qui).** Viceversa, ogni funzione $F:\mathbb R\to[0,1]$ che soddisfa le tre proprietà del teorema è la funzione di ripartizione di **qualche** variabile aleatoria: le tre proprietà **caratterizzano** completamente le funzioni di ripartizione, non sono solo condizioni necessarie.

```python
import numpy as np
import matplotlib.pyplot as plt

# Costruiamo F_X per X = somma di due dadi equi, e verifichiamo
# le tre proprieta' (non decrescenza, continuita' da destra a scalini,
# limiti 0 e 1) direttamente dalla distribuzione discreta.

esiti = [(i, j) for i in range(1, 7) for j in range(1, 7)]
somme = [i + j for i, j in esiti]
valori_possibili = sorted(set(somme))
p = {v: somme.count(v) / len(somme) for v in valori_possibili}

def F_X(x):
    return sum(prob for v, prob in p.items() if v <= x)

xs = np.linspace(1, 13, 1000)
Fs = [F_X(x) for x in xs]

fig, ax = plt.subplots(figsize=(8, 5))
ax.step(xs, Fs, where='post')
ax.set_xlabel("x")
ax.set_ylabel("F_X(x) = P(X <= x)")
ax.set_title("Funzione di ripartizione: somma di due dadi (a scalini, continua da destra)")
plt.tight_layout()
plt.show()

print("F_X non decrescente:", all(Fs[i] <= Fs[i+1] + 1e-9 for i in range(len(Fs)-1)))
print(f"F_X(0.99) = {F_X(0.99):.4f}  (atteso: 0, nessun valore <= 0.99)")
print(f"F_X(13) = {F_X(13):.4f}  (atteso: 1)")
```

## Esercizio svolto e commentato

**Problema:** sia $\Omega=[0,1]$ con $P=$ misura di Lebesgue (notebook 17.1), e $X(\omega)=\omega^2$. Verifica che $X$ è una variabile aleatoria, calcola $F_X$ esplicitamente, e verifica le tre proprietà del teorema.

**Ragionamento:**

**Passo 1 — verifichiamo la misurabilità.** $X(\omega)=\omega^2$ è una funzione polinomiale (notebook 3.11), quindi continua su $[0,1]$ (notebook 5.6: le funzioni elementari, inclusi i polinomi, sono continue nel loro dominio), e ogni funzione continua è misurabile rispetto alla σ-algebra di Borel (notebook 13.3, un fatto generale lì dimostrato): quindi $X$ è una variabile aleatoria.

**Passo 2 — calcoliamo $F_X(x)$ per $x\in[0,1]$.** $F_X(x)=P(X\le x)=P(\omega^2\le x)=P(\omega\le\sqrt x)$ (essendo $\omega\ge0$ su $[0,1]$, la disuguaglianza si inverte prendendo la radice, funzione crescente su $[0,\infty)$, notebook 3.9) $=P\big([0,\sqrt x]\big)=\sqrt x$ (misura di Lebesgue di un intervallo, notebook 13.2, essendo $\sqrt x\in[0,1]$).

**Passo 3 — completiamo $F_X$ fuori da $[0,1]$.** Per $x<0$: $\{X\le x\}=\emptyset$ (essendo $X\ge0$ sempre), quindi $F_X(x)=0$. Per $x>1$: $\{X\le x\}=\Omega$ (essendo $X\le1$ sempre), quindi $F_X(x)=1$.

**Passo 4 — verifichiamo le tre proprietà.** **Non decrescenza:** $\sqrt x$ è crescente su $[0,1]$ (notebook 3.9), e le code costanti $0,1$ raccordano senza scendere. **Continuità da destra:** $\sqrt x$ è continua ovunque su $[0,1]$ (in particolare da destra, notebook 5.6), incluso nei punti di raccordo $x=0$ ($\sqrt0=0$, coerente con la coda sinistra) e $x=1$ ($\sqrt1=1$, coerente con la coda destra) — qui $F_X$ è addirittura continua **anche da sinistra**, un fatto speciale legato al fatto che $X$ è una variabile continua (nessun "salto"), non garantito in generale (come mostra l'esempio discreto della cella di codice). **Limiti:** $\lim_{x\to-\infty}F_X(x)=0$ e $\lim_{x\to+\infty}F_X(x)=1$, immediati dalle code costanti.

```python
import sympy as sp

omega, x = sp.symbols('omega x', real=True, nonnegative=True)

# F_X(x) = P(omega^2 <= x) = P(omega <= sqrt(x)) = sqrt(x) su [0,1]
F_X = sp.sqrt(x)

# Verifica di non decrescenza: la derivata deve essere >= 0 su (0,1)
derivata = sp.diff(F_X, x)
print("F_X'(x) =", derivata, " (positiva su (0,1): F_X e' crescente)")

# Verifica dei valori ai bordi
print("F_X(0) =", F_X.subs(x, 0))
print("F_X(1) =", F_X.subs(x, 1))

# confronto numerico con simulazione Monte Carlo
import numpy as np
np.random.seed(0)
campioni = np.random.uniform(0, 1, 500000)
X_campioni = campioni**2

for x_test in [0.1, 0.25, 0.5, 0.81]:
    F_simulata = np.mean(X_campioni <= x_test)
    F_teorica = np.sqrt(x_test)
    print(f"F_X({x_test}) simulata = {F_simulata:.4f}, teorica = {F_teorica:.4f}")
```

**4.** Sia $y_n\uparrow x$ (successione crescente con limite $x$, ad esempio $y_n=x-1/n$). Gli eventi $\{y_n<X\le x\}$ formano una successione **decrescente** (essendo $y_n$ crescente, l'intervallo $(y_n,x]$ si restringe) con $\bigcap_n\{y_n<X\le x\}=\{X=x\}$ (un punto soddisfa $y_n<X(\omega)\le x$ per ogni $n$ se e solo se $X(\omega)=x$, dato che $y_n\to x$). Per l'esercizio 3, $P(y_n<X\le x)=F_X(x)-F_X(y_n)$, quindi $F_X(x)-F_X(y_n)\to P(X=x)$, cioè $P(X=x)=F_X(x)-\lim_nF_X(y_n)=F_X(x)-\lim_{y\to x^-}F_X(y)$. Poiché $F_X$ è sempre continua da destra (teoria), $F_X$ è continua in $x$ (cioè continua anche da sinistra, quindi ovunque) se e solo se $\lim_{y\to x^-}F_X(y)=F_X(x)$, cioè se e solo se $P(X=x)=0$: nessun salto in $x$. ## Riepilogo finale

- Una **variabile aleatoria** è una funzione misurabile $X:\Omega\to\mathbb R$ (notebook 13.3): la misurabilità garantisce che eventi come $\{X\le x\}$ siano legittimi.
- Ogni variabile aleatoria induce una **distribuzione** $\mu_X$, una misura di probabilità su $(\mathbb R,\mathcal B(\mathbb R))$: da qui in poi si può lavorare su $\mathbb R$ dimenticando lo spazio campionario originale.
- La **funzione di ripartizione** $F_X(x)=P(X\le x)$ è sempre non decrescente, continua da destra, con limiti $0$ e $1$ agli estremi — proprietà dimostrate usando esclusivamente le proprietà di continuità delle misure di probabilità (notebook 17.1).
- Il "salto" di $F_X$ in un punto $x$ è esattamente $P(X=x)$: $F_X$ è continua ovunque se e solo se nessun valore singolo ha probabilità positiva.
- Questo linguaggio (distribuzione, funzione di ripartizione) prepara il terreno per classificare le variabili aleatorie in famiglie standard, l'oggetto del prossimo notebook.
