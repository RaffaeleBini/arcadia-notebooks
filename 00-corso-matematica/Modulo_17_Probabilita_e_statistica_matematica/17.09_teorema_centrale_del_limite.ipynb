# 17.9 — Teorema centrale del limite

**Prerequisiti:** 17.8 Legge dei grandi numeri, 15.2 Trasformata di Fourier (rigorosa), 9.3 Formula di Eulero, 9.9 Teorema di Cauchy.


## Perché ci interessa

Il notebook 17.8 ha mostrato che $\overline X_n\to\mu$, ma non ha detto **quanto velocemente**, né **come sono distribuite** le fluttuazioni residue $\overline X_n-\mu$. Il teorema centrale del limite (CLT) risponde a entrambe le domande con una scoperta sorprendente: riscalando le fluttuazioni della quantità giusta, $\sqrt n(\overline X_n-\mu)$, la loro distribuzione **converge sempre alla stessa distribuzione normale**, indipendentemente da come erano distribuite le $X_i$ di partenza (purché abbiano varianza finita).

Lo strumento tecnico per dimostrarlo è la **funzione caratteristica**, una parente stretta della trasformata di Fourier del notebook 15.2 — applicata qui, per la prima volta in questo corso, a una misura di probabilità invece che a una funzione integrabile su $\mathbb R$.

## Teoria, passo dopo passo

### La funzione caratteristica

La **funzione caratteristica** di una variabile aleatoria $X$ è
$$\varphi_X(t)=E[e^{itX}]=\int_{\mathbb R}e^{itx}\,d\mu_X(x),\qquad t\in\mathbb R,$$
dove $e^{itx}=\cos(tx)+i\sin(tx)$ (formula di Eulero, notebook 9.3). L'integrale è sempre ben definito e finito, per **ogni** distribuzione: $|e^{itx}|=1$ (essendo $\cos^2+\sin^2=1$), quindi $E[|e^{itX}|]=E[1]=1<\infty$ — a differenza del valore atteso ordinario, $\varphi_X$ esiste sempre, anche per distribuzioni senza momenti finiti (come la Cauchy del notebook 17.4, esercizio 3).

**Relazione con la trasformata di Fourier (notebook 15.2).** Se $X$ ha densità $f$, allora $\varphi_X(t)=\displaystyle\int_{\mathbb R}f(x)e^{itx}\,dx$. Confrontando con la definizione del notebook 15.2, $\hat f(\xi)=\int f(x)e^{-2\pi ix\xi}\,dx$: ponendo $\xi=-t/(2\pi)$, si ottiene $\hat f(-t/(2\pi))=\int f(x)e^{itx}\,dx=\varphi_X(t)$. La funzione caratteristica **è**, a meno di questo cambio di convenzione (nessun fattore $2\pi$, segno opposto all'esponente), la trasformata di Fourier della densità — non una nuova teoria, ma la stessa costruzione del notebook 15.2, riscalata per le convenzioni tipiche della probabilità.

**Proprietà elementari.**
- $\varphi_X(0)=E[1]=1$, $|\varphi_X(t)|\le1$ per ogni $t$ (dalla disuguaglianza $|E[Z]|\le E[|Z|]$, valida per l'integrale di Lebesgue, applicata a $Z=e^{itX}$).
- $\varphi_X$ è **continua** su tutto $\mathbb R$: se $t_n\to t$, $e^{it_nX}\to e^{itX}$ puntualmente e $|e^{it_nX}|=1$ è dominato dalla costante $1\in L^1(\Omega,P)$, quindi per il teorema di convergenza dominata (notebook 13.7) $\varphi_X(t_n)=E[e^{it_nX}]\to E[e^{itX}]=\varphi_X(t)$ — lo stesso argomento usato nel notebook 15.2 per la continuità della trasformata di Fourier.
- **Fattorizzazione per indipendenti.** Se $X,Y$ sono indipendenti (notebook 17.6): $\varphi_{X+Y}(t)=E[e^{it(X+Y)}]=E[e^{itX}e^{itY}]=E[e^{itX}]E[e^{itY}]=\varphi_X(t)\varphi_Y(t)$ (usando $E[UV]=E[U]E[V]$ per indipendenti, notebook 17.6, applicato a $U=e^{itX},V=e^{itY}$, entrambe funzioni misurabili di $X$ e $Y$ rispettivamente, quindi indipendenti per il notebook 17.6 esercizio 2). Questa è la proprietà chiave: **somme di indipendenti diventano prodotti di funzioni caratteristiche**, esattamente come la trasformata di Fourier trasforma convoluzioni in prodotti.

### Il teorema di continuità di Lévy (enunciato)

**Teorema (Lévy, senza dimostrazione completa).** Se $\varphi_{X_n}(t)\to\varphi(t)$ per ogni $t$, e $\varphi$ è continua in $t=0$, allora $\varphi$ è la funzione caratteristica di una variabile aleatoria $X$, e $X_n\to X$ in distribuzione (notebook 17.7).

Questo teorema (la cui dimostrazione userebbe un argomento di inversione simile, nello spirito, al teorema di inversione di Fourier del notebook 15.2, ma applicato a misure invece che a funzioni $L^1$) è il ponte che ci serve: dimostreremo la convergenza delle funzioni caratteristiche, un calcolo esplicito, e questo teorema tradurrà automaticamente il risultato in convergenza in distribuzione.

### Il teorema centrale del limite

**Teorema (Lindeberg-Lévy).** Siano $X_1,X_2,\ldots$ i.i.d. con $E[X_i]=\mu$, $\mathrm{Var}(X_i)=\sigma^2\in(0,\infty)$. Allora
$$Z_n:=\frac{\overline X_n-\mu}{\sigma/\sqrt n}=\frac{\sum_{i=1}^n(X_i-\mu)}{\sigma\sqrt n}\ \xrightarrow{d}\ \mathrm N(0,1).$$

**Dimostrazione (schema, tramite funzioni caratteristiche).** Poniamo $Y_i=(X_i-\mu)/\sigma$ (standardizzate: $E[Y_i]=0$, $\mathrm{Var}(Y_i)=1$), così $Z_n=\dfrac1{\sqrt n}\sum_{i=1}^nY_i$. Sia $\psi(t)=\varphi_{Y_1}(t)$ (la stessa per ogni $i$, essendo le $X_i$ identicamente distribuite). Per fattorizzazione (le $Y_i$ sono indipendenti, essendo funzioni delle $X_i$ indipendenti) e la proprietà $\varphi_{aX}(t)=\varphi_X(at)$ (immediata dalla definizione, sostituendo $itaX$ al posto di $itX$):
$$\varphi_{Z_n}(t)=\varphi_{\sum_iY_i}\Big(\frac t{\sqrt n}\Big)=\Big[\psi\Big(\frac t{\sqrt n}\Big)\Big]^n.$$

Espandiamo $\psi$ vicino a $0$ con la formula di Taylor (notebook 5.14, applicata separatamente a parte reale e immaginaria, entrambe funzioni lisce se $E[Y_i^2]<\infty$): $\psi(s)=\psi(0)+\psi'(0)s+\frac12\psi''(0)s^2+o(s^2)$. Calcolando le derivate (derivando sotto il segno di integrale, giustificato dalla dominazione uniforme che $E[Y_i^2]<\infty$ garantisce): $\psi(0)=1$, $\psi'(0)=iE[Y_1]=0$, $\psi''(0)=i^2E[Y_1^2]=-1$ (usando $E[Y_1^2]=\mathrm{Var}(Y_1)+E[Y_1]^2=1+0=1$). Quindi $\psi(s)=1-\dfrac{s^2}2+o(s^2)$ per $s\to0$.

Sostituendo $s=t/\sqrt n$:
$$\varphi_{Z_n}(t)=\Big[1-\frac{t^2}{2n}+o\Big(\frac1n\Big)\Big]^n\ \xrightarrow[n\to\infty]{}\ e^{-t^2/2}$$
(usando il limite notevole $(1+1/n)^n\to e$ del notebook 5.5, esteso con un argomento di continuità al caso $(1+a_n/n)^n\to e^a$ quando $a_n\to a$, con $a_n=-t^2/2+o(1)\to-t^2/2$). $\blacksquare$ (a meno del teorema di Lévy, che traduce questo limite puntuale delle funzioni caratteristiche in convergenza in distribuzione).

### Identificazione del limite: $e^{-t^2/2}$ è la funzione caratteristica della normale standard

Resta da verificare che $e^{-t^2/2}$ sia davvero $\varphi_Z(t)$ per $Z\sim\mathrm N(0,1)$, altrimenti l'identificazione del limite con la normale standard resterebbe incompleta.

$$\varphi_Z(t)=\int_{\mathbb R}e^{itx}\frac1{\sqrt{2\pi}}e^{-x^2/2}\,dx=\frac1{\sqrt{2\pi}}\int_{\mathbb R}e^{-x^2/2+itx}\,dx.$$

Completiamo il quadrato nell'esponente: $-\frac{x^2}2+itx=-\frac12(x-it)^2-\frac{t^2}2$ (verifica diretta: $-\frac12(x-it)^2=-\frac12x^2+itx+\frac12i^2t^2=-\frac{x^2}2+itx-\frac{t^2}2$, quindi sommando $t^2/2$ si ottiene esattamente $-\frac{x^2}2+itx$). Quindi:
$$\varphi_Z(t)=\frac{e^{-t^2/2}}{\sqrt{2\pi}}\int_{\mathbb R}e^{-\frac12(x-it)^2}\,dx.$$

L'integrale rimanente è un integrale gaussiano lungo la retta orizzontale $\{\mathrm{Im}(z)=-t\}$ nel piano complesso (ponendo $z=x-it$). Per il **teorema di Cauchy** (notebook 9.9), l'integrale di una funzione olomorfa (qui $e^{-z^2/2}$, olomorfa su tutto $\mathbb C$, notebook 9.6) lungo il bordo di un rettangolo è zero; applicandolo a un rettangolo con lati orizzontali a $\mathrm{Im}=0$ e $\mathrm{Im}=-t$ e lati verticali che, facendoli tendere a $\pm\infty$, danno contributo nullo (essendo $|e^{-z^2/2}|$ trascurabile per $|\mathrm{Re}(z)|$ grande, a $\mathrm{Im}(z)$ fissata), si ottiene che l'integrale sulla retta orizzontale a quota $-t$ **coincide** con quello a quota $0$: $\displaystyle\int_{\mathbb R}e^{-\frac12(x-it)^2}\,dx=\int_{\mathbb R}e^{-x^2/2}\,dx=\sqrt{2\pi}$ (l'integrale gaussiano standard, conseguenza diretta di $\int e^{-y^2}dy=\sqrt\pi$ del notebook 7.15, con $y=x/\sqrt2$). Quindi
$$\varphi_Z(t)=\frac{e^{-t^2/2}}{\sqrt{2\pi}}\cdot\sqrt{2\pi}=e^{-t^2/2}$$
— esattamente il limite trovato sopra: il CLT converge davvero alla normale standard.

```python
import numpy as np
import matplotlib.pyplot as plt

# Verifica visiva del CLT: sommiamo n variabili Bernoulli(0.3) (fortemente
# NON gaussiane per n piccolo), standardizziamo, e confrontiamo l'istogramma
# con la densita' normale standard, per diversi n.

np.random.seed(7)
p = 0.3
mu, sigma2 = p, p*(1-p)
sigma = np.sqrt(sigma2)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, n in zip(axes, [1, 10, 50]):
    n_simulazioni = 100000
    somme = np.random.binomial(n, p, size=n_simulazioni)
    X_barra = somme / n
    Z_n = (X_barra - mu) / (sigma / np.sqrt(n))

    ax.hist(Z_n, bins=50, density=True, alpha=0.6, label=f'Z_n (n={n})')
    x_gauss = np.linspace(-4, 4, 200)
    ax.plot(x_gauss, np.exp(-x_gauss**2/2)/np.sqrt(2*np.pi), 'r-', label='N(0,1)')
    ax.set_title(f"n={n}")
    ax.legend(fontsize=8)
    ax.set_xlim(-4, 4)
plt.tight_layout()
plt.show()

print("Anche partendo da una Bernoulli (estremamente asimmetrica per n piccolo),")
print("la distribuzione di Z_n si avvicina alla normale standard al crescere di n.")
```

## Esercizio svolto e commentato

**Problema:** verifica numericamente la convergenza $\varphi_{Z_n}(t)\to e^{-t^2/2}$ per $t=1$, usando $X_i\sim\mathrm{Exp}(1)$ (quindi $\mu=1,\sigma^2=1$), calcolando $\varphi_{Z_n}(1)$ per $n$ crescente tramite simulazione Monte Carlo.

**Ragionamento:**

**Passo 1 — richiamiamo la definizione.** $\varphi_{Z_n}(t)=E[e^{itZ_n}]=E[\cos(tZ_n)]+iE[\sin(tZ_n)]$ (parte reale e immaginaria, formula di Eulero, notebook 9.3), dove $Z_n=\dfrac{\sum_{i=1}^n(X_i-1)}{\sqrt n}$ (usando $\mu=\sigma=1$ per $\mathrm{Exp}(1)$).

**Passo 2 — stimiamo $E[\cos(tZ_n)]$ e $E[\sin(tZ_n)]$ con Monte Carlo (metodo giustificato dalla legge dei grandi numeri, notebook 17.8).** Generiamo molte repliche indipendenti di $Z_n$ (ciascuna richiede $n$ campioni esponenziali), e calcoliamo le medie campionarie di $\cos(tZ_n)$ e $\sin(tZ_n)$.

**Passo 3 — confrontiamo con il valore limite.** Per $t=1$: $e^{-1^2/2}=e^{-0.5}\approx0.6065$. Ci aspettiamo che la parte reale stimata si avvicini a questo valore al crescere di $n$, e la parte immaginaria si avvicini a $0$ (essendo $e^{-t^2/2}$ un numero reale, senza parte immaginaria).

**Conclusione (da verificare nella cella seguente):** la convergenza empirica di $\varphi_{Z_n}(1)$ verso $e^{-0.5}$ conferma concretamente, su un esempio con distribuzione di partenza fortemente asimmetrica (l'esponenziale non è affatto simmetrica attorno alla propria media, a differenza della normale limite), il calcolo formale svolto nella teoria.

```python
import numpy as np

def phi_Z_n_stimata(n, t, n_ripetizioni=200000):
    campioni = np.random.exponential(scale=1.0, size=(n_ripetizioni, n))
    X_barra = campioni.mean(axis=1)
    Z_n = (X_barra - 1.0) / (1.0 / np.sqrt(n))  # mu=sigma=1 per Exp(1)
    parte_reale = np.mean(np.cos(t * Z_n))
    parte_immaginaria = np.mean(np.sin(t * Z_n))
    return parte_reale, parte_immaginaria

t = 1.0
valore_limite = np.exp(-t**2 / 2)
print(f"Valore limite atteso: e^(-0.5) = {valore_limite:.4f}\n")

np.random.seed(8)
for n in [1, 5, 20, 100]:
    re, im = phi_Z_n_stimata(n, t)
    print(f"n={n:4d}: phi_Z_n(1) stimata = {re:.4f} + {im:.4f}i")
```

## Riepilogo finale

- La **funzione caratteristica** $\varphi_X(t)=E[e^{itX}]$, una parente della trasformata di Fourier (notebook 15.2) applicata a misure di probabilità, esiste sempre e trasforma somme di indipendenti in prodotti.
- Il **teorema centrale del limite** afferma che $\sqrt n(\overline X_n-\mu)/\sigma\to\mathrm N(0,1)$ in distribuzione, per **qualunque** distribuzione di partenza con varianza finita.
- La dimostrazione (tramite espansione di Taylor della funzione caratteristica standardizzata e il teorema di continuità di Lévy) usa solo il fatto che i primi due momenti determinano $\psi$ al secondo ordine vicino a $0$: da qui l'universalità del limite normale.
- Il completamento del quadrato nell'esponente, insieme al teorema di Cauchy (notebook 9.9) per giustificare lo spostamento del cammino di integrazione nel piano complesso, identifica rigorosamente $e^{-t^2/2}$ come la funzione caratteristica della normale standard.
- L'ipotesi di varianza finita è essenziale: senza di essa, il termine quadratico dell'espansione di Taylor non esiste, e il limite (quando esiste) non è più gaussiano.
