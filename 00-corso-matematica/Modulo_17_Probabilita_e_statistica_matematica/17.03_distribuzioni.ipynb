## Teoria, passo dopo passo

### Variabili aleatorie discrete

$X$ è **discreta** se assume valori in un insieme finito o numerabile $\{x_1,x_2,\ldots\}$. La sua **funzione di massa di probabilità** (PMF) è $p_X(x_i)=P(X=x_i)$, con $\sum_ip_X(x_i)=1$ (σ-additività della probabilità, notebook 17.1, applicata alla partizione $\Omega=\bigsqcup_i\{X=x_i\}$). La distribuzione $\mu_X$ è determinata da $p_X$: $\mu_X(B)=\sum_{x_i\in B}p_X(x_i)$.

**Famiglie discrete standard:**
- **Bernoulli** $\mathrm{Be}(p)$: $X\in\{0,1\}$, $P(X=1)=p$, $P(X=0)=1-p$ — già incontrata nel notebook 17.2, esercizio 1.
- **Binomiale** $\mathrm{Bin}(n,p)$: numero di successi in $n$ prove Bernoulli indipendenti, $P(X=k)=\binom nkp^k(1-p)^{n-k}$ per $k=0,\ldots,n$, dove $\binom nk$ conta il numero di modi di scegliere quali $k$ prove, tra le $n$, sono successi — il **coefficiente binomiale**, che il Modulo 18 tratterà sistematicamente con gli strumenti del calcolo combinatorio.
- **Poisson** $\mathrm{Poi}(\lambda)$: $P(X=k)=\dfrac{\lambda^k}{k!}e^{-\lambda}$ per $k=0,1,2,\ldots$, il limite della binomiale quando $n\to\infty,p\to0$ con $np\to\lambda$ (un fatto che l'esercizio svolto verifica numericamente).

### Variabili aleatorie continue e densità

$X$ è **(assolutamente) continua** se esiste una funzione $f_X:\mathbb R\to[0,\infty)$, misurabile (notebook 13.3), tale che
$$F_X(x)=\int_{-\infty}^xf_X(t)\,dt\qquad\text{per ogni }x,$$
dove l'integrale è quello di Lebesgue (notebook 13.4). La funzione $f_X$ si chiama **densità** di $X$. Più in generale, per ogni boreliano $B$: $\mu_X(B)=\int_Bf_X\,d\lambda$ (una formula che, per $B=(-\infty,x]$, si riduce a quella data sopra; la verifica che questa assegnazione definisca effettivamente una misura è analoga, punto per punto, a quella già vista per l'integrale di Lebesgue rispetto a una densità, notebook 13.4). Necessariamente $\int_{\mathbb R}f_X\,d\lambda=F_X(+\infty)=1$ (notebook 17.2).

**Nota (fuori dallo scopo di questo corso).** Non ogni distribuzione con $F_X$ continua ha una densità in questo senso (esistono controesempi patologici, le cosiddette distribuzioni "singolari continue"). Il fatto generale che identifica esattamente quali misure ammettono una densità rispetto a $\lambda$ va sotto il nome di **teorema di Radon-Nikodym**, che non dimostriamo in questo corso: qui ci limitiamo a lavorare con densità date esplicitamente.

**Perché una densità non è una probabilità.** Un errore comune: $f_X(x)$ **non** è $P(X=x)$ (che vale sempre $0$ per una variabile continua, notebook 17.2 esercizio 4). $f_X(x)$ è piuttosto una "densità di probabilità per unità di lunghezza": $P(X\in[x,x+dx])\approx f_X(x)\,dx$ per $dx$ infinitesimo, e solo integrando su un intervallo si ottiene una vera probabilità.

**Famiglie continue standard:**
- **Uniforme** $\mathrm U(a,b)$: $f_X(x)=\dfrac1{b-a}$ per $x\in[a,b]$, $0$ altrove — la generalizzazione diretta dell'esempio "punto a caso in $[0,1]$" del notebook 17.1 (che è $\mathrm U(0,1)$).
- **Esponenziale** $\mathrm{Exp}(\lambda)$: $f_X(x)=\lambda e^{-\lambda x}$ per $x\ge0$, $0$ altrove (verifica diretta, notebook 5.17: $\int_0^\infty\lambda e^{-\lambda x}\,dx=1$, un integrale improprio, notebook 5.18).
- **Normale (o gaussiana)** $\mathrm N(\mu,\sigma^2)$: $f_X(x)=\dfrac1{\sigma\sqrt{2\pi}}e^{-(x-\mu)^2/(2\sigma^2)}$. La normalizzazione $\int_{\mathbb R}f_X\,dx=1$ segue dall'integrale gaussiano $\int_{\mathbb R}e^{-y^2}\,dy=\sqrt\pi$, dimostrato nel notebook 7.15 con il trucco delle coordinate polari, tramite il cambio di variabile $y=(x-\mu)/(\sigma\sqrt2)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import comb

# Verifica: la Binomiale(n,p) converge alla Poisson(lambda) quando n->infinito,
# p->0, con n*p = lambda fissato.

lam = 3.0
k_vals = np.arange(0, 15)

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)
for ax, n in zip(axes, [20, 100, 1000]):
    p = lam / n
    pmf_binomiale = [comb(n, k) * p**k * (1-p)**(n-k) for k in k_vals]
    pmf_poisson = [lam**k / np.math.factorial(k) * np.exp(-lam) for k in k_vals]
    ax.bar(k_vals - 0.2, pmf_binomiale, width=0.4, label=f'Bin(n={n}, p={p:.4f})')
    ax.bar(k_vals + 0.2, pmf_poisson, width=0.4, label=f'Poisson(lambda={lam})')
    ax.legend(fontsize=8)
    ax.set_title(f"n={n}")
plt.tight_layout()
plt.show()

errore_max = max(abs(comb(1000, k) * (lam/1000)**k * (1-lam/1000)**(1000-k)
                     - lam**k/np.math.factorial(k)*np.exp(-lam)) for k in k_vals)
print(f"Errore massimo (n=1000) tra Binomiale e Poisson: {errore_max:.6f}")
```

## Esercizio svolto e commentato

**Problema:** verifica che $f_X(x)=\dfrac1{\sigma\sqrt{2\pi}}e^{-(x-\mu)^2/(2\sigma^2)}$ è effettivamente una densità (cioè $\int_{\mathbb R}f_X\,dx=1$), usando l'integrale gaussiano del notebook 7.15.

**Ragionamento:**

**Passo 1 — richiamiamo il fatto noto.** Nel notebook 7.15, usando le coordinate polari per calcolare $\Big(\int_{\mathbb R}e^{-y^2}dy\Big)^2$ come un integrale doppio su tutto il piano, si dimostra $\int_{\mathbb R}e^{-y^2}\,dy=\sqrt\pi$.

**Passo 2 — cambiamo variabile per adattare l'integrale al nostro caso.** Poniamo $y=\dfrac{x-\mu}{\sigma\sqrt2}$, quindi $x=\mu+\sigma\sqrt2\,y$, $dx=\sigma\sqrt2\,dy$ (cambiamento di variabile, notebook 7.15), e $(x-\mu)^2/(2\sigma^2)=y^2$. L'integrale diventa:
$$\int_{\mathbb R}e^{-(x-\mu)^2/(2\sigma^2)}\,dx=\int_{\mathbb R}e^{-y^2}\cdot\sigma\sqrt2\,dy=\sigma\sqrt2\int_{\mathbb R}e^{-y^2}\,dy=\sigma\sqrt2\cdot\sqrt\pi=\sigma\sqrt{2\pi}.$$

**Passo 3 — dividiamo per la costante di normalizzazione.** $\displaystyle\int_{\mathbb R}f_X(x)\,dx=\frac1{\sigma\sqrt{2\pi}}\int_{\mathbb R}e^{-(x-\mu)^2/(2\sigma^2)}\,dx=\frac1{\sigma\sqrt{2\pi}}\cdot\sigma\sqrt{2\pi}=1$.

**Conclusione:** $f_X$ è effettivamente una densità di probabilità, per ogni scelta di $\mu\in\mathbb R$ e $\sigma>0$ — la costante $\frac1{\sigma\sqrt{2\pi}}$ davanti all'esponenziale non è arbitraria: è **esattamente** ciò che serve per compensare il fattore $\sigma\sqrt{2\pi}$ prodotto dall'integrale gaussiano, garantendo che l'area totale sotto la curva sia $1$.

```python
import sympy as sp

x, mu, sigma, y = sp.symbols('x mu sigma y', real=True)
sigma_pos = sp.symbols('sigma', positive=True)

f_X = (1/(sigma_pos*sp.sqrt(2*sp.pi))) * sp.exp(-(x-mu)**2/(2*sigma_pos**2))

integrale = sp.integrate(f_X, (x, -sp.oo, sp.oo))
print("Integrale di f_X su tutto R:", sp.simplify(integrale))

# Verifica numerica per valori specifici
f_num = f_X.subs({mu: 2, sigma_pos: 1.5})
from scipy import integrate as spint
import numpy as np
f_lambda = sp.lambdify(x, f_num, 'numpy')
valore, _ = spint.quad(f_lambda, -20, 20)
print(f"Integrale numerico (mu=2, sigma=1.5): {valore:.6f}")
```

**1.** $\int_0^\infty\lambda e^{-\lambda x}\,dx=\lambda\lim_{M\to\infty}\Big[-\frac1\lambda e^{-\lambda x}\Big]_0^M=\lambda\lim_{M\to\infty}\Big(-\frac1\lambda e^{-\lambda M}+\frac1\lambda\Big)=\lambda\cdot\frac1\lambda=1$ (essendo $e^{-\lambda M}\to0$ per $M\to\infty$, $\lambda>0$, notebook 5.18): $f_X$ è una densità valida.

**2.** Per $x<a$: $F_X(x)=0$ (nessuna massa a sinistra di $a$). Per $x\in[a,b]$: $F_X(x)=\int_a^x\frac1{b-a}\,dt=\frac{x-a}{b-a}$. Per $x>b$: $F_X(x)=1$. Verifica: **non decrescente** ($\frac{x-a}{b-a}$ cresce linearmente su $[a,b]$, le code sono costanti); **continua da destra** ($F_X$ qui è continua ovunque, essendo lineare a tratti e raccordata senza salti: $F_X(a)=0$ da entrambi i lati, $F_X(b)=1$ da entrambi i lati); **limiti** $0$ e $1$ dalle code costanti.

**3.** $P(X>s+t\mid X>s)=\dfrac{P(X>s+t,X>s)}{P(X>s)}=\dfrac{P(X>s+t)}{P(X>s)}$ (essendo $s+t\ge s$, l'evento $\{X>s+t\}\subseteq\{X>s\}$, quindi la loro intersezione è $\{X>s+t\}$). $P(X>u)=1-F_X(u)=1-(1-e^{-\lambda u})=e^{-\lambda u}$ (calcolando $F_X$ per l'esponenziale in modo analogo all'esercizio 1). Quindi $P(X>s+t\mid X>s)=\dfrac{e^{-\lambda(s+t)}}{e^{-\lambda s}}=e^{-\lambda t}=P(X>t)$: **l'esponenziale "dimentica" quanto tempo è già trascorso** — una proprietà che la rende il modello standard per tempi di attesa "senza usura" (decadimento radioattivo, tempo di guasto di componenti senza logorio).

**4.** $\int_{\mathbb R}xf_X(x)\,dx=\frac1{\sigma\sqrt{2\pi}}\int_{\mathbb R}xe^{-(x-\mu)^2/(2\sigma^2)}\,dx$. Con $y=(x-\mu)/(\sigma\sqrt2)$, $x=\mu+\sigma\sqrt2y$, $dx=\sigma\sqrt2\,dy$ (esercizio svolto): $=\frac1{\sigma\sqrt{2\pi}}\int_{\mathbb R}(\mu+\sigma\sqrt2y)e^{-y^2}\sigma\sqrt2\,dy=\frac1{\sqrt\pi}\Big[\mu\int_{\mathbb R}e^{-y^2}dy+\sigma\sqrt2\int_{\mathbb R}ye^{-y^2}dy\Big]$. Il primo integrale è $\sqrt\pi$ (esercizio svolto); il secondo è **nullo** perché $ye^{-y^2}$ è una funzione **dispari** (notebook 3.8) integrata su un dominio simmetrico $\mathbb R$ (l'area a sinistra di $0$ cancella esattamente quella a destra). Quindi $\int_{\mathbb R}xf_X(x)\,dx=\frac1{\sqrt\pi}\big[\mu\sqrt\pi+0\big]=\mu$: il parametro $\mu$ è, come suggerisce il nome, il **valore atteso** della normale — un fatto che dimostreremo in generale nel prossimo notebook.

## Riepilogo finale

- Le variabili **discrete** hanno una PMF $p_X$ che assegna massa positiva a un insieme numerabile di punti; le variabili **continue** hanno una densità $f_X$, con $F_X(x)=\int_{-\infty}^xf_X\,d\lambda$ (integrale di Lebesgue, notebook 13.4).
- Una densità **non** è una probabilità puntuale: va sempre integrata su un insieme per dare una probabilità.
- Le famiglie standard (Bernoulli, Binomiale, Poisson tra le discrete; Uniforme, Esponenziale, Normale tra le continue) coprono la stragrande maggioranza dei modelli usati in pratica, e la Poisson emerge come limite della Binomiale per eventi rari.
- La costante di normalizzazione della densità normale è determinata esattamente dall'integrale gaussiano (notebook 7.15): nessun'altra scelta renderebbe $f_X$ una densità valida.
- L'esponenziale ha la proprietà di assenza di memoria, unica tra le distribuzioni continue positive.
