## Teoria, passo dopo passo

### Lo spazio di probabilità

Uno **spazio di probabilità** è una tripla $(\Omega,\mathcal F,P)$, dove:
- $\Omega$ è l'insieme di tutti gli esiti possibili di un esperimento (lo **spazio campionario**);
- $\mathcal F$ è una **σ-algebra** su $\Omega$ (notebook 13.1): i suoi elementi si chiamano **eventi**, gli unici sottoinsiemi di $\Omega$ a cui ha senso assegnare una probabilità;
- $P:\mathcal F\to[0,1]$ è una **misura** (notebook 13.2: $P(\emptyset)=0$, σ-additività su unioni numerabili disgiunte) che soddisfa la condizione aggiuntiva di **normalizzazione**: $P(\Omega)=1$.

Questi tre ingredienti sono i cosiddetti **assiomi di Kolmogorov**: non sono nuovi assiomi indipendenti, ma esattamente gli assiomi di misura del notebook 13.2, più $P(\Omega)=1$. Un evento $A\in\mathcal F$ **si verifica** se l'esito effettivo dell'esperimento appartiene ad $A$; $P(A)$ è la probabilità che questo accada.

### Proprietà elementari, ereditate dal Modulo 13

Tutte le proprietà delle misure dimostrate nel notebook 13.2 valgono, senza bisogno di ridimostrarle, per $P$:

- **Monotonia:** se $A\subseteq B$, $P(A)\le P(B)$ (notebook 13.2).
- **Probabilità del complementare:** $P(A^c)=1-P(A)$, conseguenza immediata dell'additività applicata a $\Omega=A\sqcup A^c$ e della normalizzazione $P(\Omega)=1$.
- **Inclusione-esclusione per due eventi:** $P(A\cup B)=P(A)+P(B)-P(A\cap B)$ (notebook 13.2, esercizio 1: la formula di inclusione-esclusione finita per due insiemi, dimostrata lì per misure generiche).
- **Continuità dal basso:** se $A_1\subseteq A_2\subseteq\cdots$, $P\big(\bigcup_nA_n\big)=\lim_nP(A_n)$ (notebook 13.2).
- **Continuità dall'alto:** se $A_1\supseteq A_2\supseteq\cdots$, $P\big(\bigcap_nA_n\big)=\lim_nP(A_n)$ — qui **senza** bisogno dell'ipotesi $P(A_1)<\infty$ richiesta nel caso generale del notebook 13.2 (esercizio 2), perché $P(A_1)\le P(\Omega)=1<\infty$ **sempre**: la normalizzazione rende la continuità dall'alto **automatica** per ogni successione decrescente di eventi, un piccolo ma utile guadagno rispetto alla teoria della misura generale.

### La σ-algebra degli eventi: perché serve, e perché di solito è quella di Borel

Se $\Omega$ è **finito o numerabile**, quasi sempre si prende $\mathcal F=\mathcal P(\Omega)$ (l'insieme delle parti, esso stesso una σ-algebra, notebook 13.1): ogni sottoinsieme è un evento legittimo. Se $\Omega=\mathbb R$ (ad esempio, l'esito di una misura fisica continua), si prende $\mathcal F=\mathcal B(\mathbb R)$, la **σ-algebra di Borel** (notebook 13.1): non tutti i sottoinsiemi di $\mathbb R$ sono misurabili in modo coerente con l'additività numerabile (un fatto profondo della teoria della misura, oltre lo scopo di questo corso), ma la classe di Borel è già abbastanza ricca da contenere ogni insieme che si incontra nella pratica.

### Esempi

- **Moneta equa:** $\Omega=\{T,C\}$, $\mathcal F=\mathcal P(\Omega)$, $P(\{T\})=P(\{C\})=1/2$.
- **Dado a sei facce:** $\Omega=\{1,\ldots,6\}$, $P(\{k\})=1/6$ per ogni $k$: la **misura del conteggio** (notebook 13.2) normalizzata dividendo per $|\Omega|=6$.
- **Punto scelto "a caso" in $[0,1]$:** $\Omega=[0,1]$, $\mathcal F=\mathcal B([0,1])$, $P(A)=\lambda(A)$, la **misura di Lebesgue** ristretta a $[0,1]$ (notebook 13.2): essendo $\lambda([0,1])=1$, questa è già una probabilità, senza bisogno di normalizzare. Questo è il primo esempio, in questo corso, di uno spazio di probabilità **non numerabile** — reso possibile esattamente dagli strumenti del Modulo 13, non disponibili prima.

```python
import numpy as np

# Verifica sperimentale (simulazione Monte Carlo) delle proprieta' elementari
# per lo spazio di probabilita' del dado, confrontando frequenze simulate con
# le probabilita' teoriche P({k}) = 1/6.

np.random.seed(0)
n_lanci = 200000
lanci = np.random.randint(1, 7, size=n_lanci)

print("Frequenze simulate (dado equo):")
for k in range(1, 7):
    freq = np.mean(lanci == k)
    print(f"  P({{{k}}}) simulata = {freq:.4f}  (teorica: {1/6:.4f})")

# Verifica di P(A union B) = P(A) + P(B) - P(A intersect B)
A = np.isin(lanci, [1, 2, 3])       # evento "esce 1, 2 o 3"
B = np.isin(lanci, [2, 4, 6])       # evento "esce pari"
P_A = np.mean(A)
P_B = np.mean(B)
P_A_and_B = np.mean(A & B)
P_A_or_B_simulata = np.mean(A | B)
P_A_or_B_formula = P_A + P_B - P_A_and_B

print(f"\nP(A union B) simulata:  {P_A_or_B_simulata:.4f}")
print(f"P(A) + P(B) - P(A intersect B): {P_A_or_B_formula:.4f}")
print(f"Coincidono (inclusione-esclusione, notebook 13.2 esercizio 1): {np.isclose(P_A_or_B_simulata, P_A_or_B_formula, atol=0.01)}")
```

## Esercizio svolto e commentato

**Problema:** in uno spazio di probabilità $(\Omega,\mathcal F,P)$, dimostra che per ogni successione di eventi $A_1,A_2,\ldots$ (non necessariamente crescente o decrescente), $P\Big(\bigcup_nA_n\Big)\le\sum_nP(A_n)$ (la **subadditività numerabile**), usando solo gli assiomi di misura.

**Ragionamento:**

**Passo 1 — il problema con l'additività diretta.** L'assioma di misura (notebook 13.2) dà l'uguaglianza $P\big(\bigcup_nA_n\big)=\sum_nP(A_n)$ solo se gli $A_n$ sono **disgiunti**. Qui non lo sono in generale: dobbiamo prima "disgiungerli" senza cambiare l'unione.

**Passo 2 — costruiamo una successione disgiunta con la stessa unione.** Poniamo $B_1=A_1$ e, per $n\ge2$, $B_n=A_n\setminus(A_1\cup\cdots\cup A_{n-1})$: per costruzione $B_n\subseteq A_n$ e i $B_n$ sono a due a due disgiunti (se $m<n$, $B_n$ esclude per definizione tutto ciò che sta in $A_m\subseteq A_1\cup\cdots\cup A_{n-1}$). Inoltre $\bigcup_nB_n=\bigcup_nA_n$ (ogni punto della seconda unione appartiene a un primo $A_n$ che lo contiene, quindi al corrispondente $B_n$).

**Passo 3 — applichiamo l'additività numerabile ai $B_n$.** $P\Big(\bigcup_nA_n\Big)=P\Big(\bigcup_nB_n\Big)=\sum_nP(B_n)$ (additività numerabile, essendo i $B_n$ disgiunti).

**Passo 4 — usiamo la monotonia.** Essendo $B_n\subseteq A_n$, per monotonia (notebook 13.2) $P(B_n)\le P(A_n)$ per ogni $n$. Sommando: $\sum_nP(B_n)\le\sum_nP(A_n)$.

**Conclusione:** combinando i passi 3 e 4, $P\Big(\bigcup_nA_n\Big)=\sum_nP(B_n)\le\sum_nP(A_n)$: la subadditività numerabile, dimostrata usando solo gli assiomi di base — nessuna proprietà "nuova" della probabilità, solo una conseguenza combinatoria degli assiomi di misura già visti nel Modulo 13.

```python
import numpy as np

# Verifica numerica della subadditivita' numerabile su un esempio concreto:
# Omega = {1,...,6} (dado), A_n = {n} per n=1..6 (eventi NON disgiunti se
# li ripetiamo, quindi costruiamo un esempio con sovrapposizioni vere)

# Esempio con sovrapposizioni: A1={1,2,3}, A2={2,3,4}, A3={3,4,5}, A4={4,5,6}
eventi = [
    {1, 2, 3},
    {2, 3, 4},
    {3, 4, 5},
    {4, 5, 6},
]
P_singolo = 1/6

somma_probabilita = sum(len(A) * P_singolo for A in eventi)
unione = set().union(*eventi)
P_unione = len(unione) * P_singolo

print(f"P(unione) = {P_unione:.4f}")
print(f"Somma delle P(A_n) = {somma_probabilita:.4f}")
print(f"Subadditivita' verificata (P(unione) <= somma): {P_unione <= somma_probabilita + 1e-9}")
```

## Riepilogo finale

- Uno **spazio di probabilità** $(\Omega,\mathcal F,P)$ è uno spazio di misura (notebook 13.1-13.2) con $P(\Omega)=1$: nessun assioma nuovo, solo un caso particolare con interpretazione nuova.
- Tutte le proprietà delle misure (monotonia, subadditività, continuità dal basso e dall'alto) valgono automaticamente per $P$; la normalizzazione rende la continuità dall'alto **sempre** valida, senza l'ipotesi tecnica richiesta nel caso generale.
- $\mathcal F$ è tipicamente $\mathcal P(\Omega)$ per spazi discreti, o la σ-algebra di Borel $\mathcal B(\mathbb R)$ (notebook 13.1) per spazi continui.
- La misura di Lebesgue normalizzata fornisce il primo esempio non banale di spazio di probabilità su un insieme non numerabile: "scegliere un punto a caso in $[0,1]$" ora ha un significato matematico rigoroso.
- Il lemma di Borel-Cantelli, dimostrato usando solo subadditività e continuità dall'alto, sarà uno strumento chiave nel notebook 17.8 per discutere la convergenza quasi certa.
