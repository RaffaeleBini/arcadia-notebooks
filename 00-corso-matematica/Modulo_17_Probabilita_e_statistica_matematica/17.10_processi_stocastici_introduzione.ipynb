## Teoria, passo dopo passo

### Definizione

Un **processo stocastico** a tempo discreto è una famiglia $(X_n)_{n\ge0}$ di variabili aleatorie, tutte definite sullo stesso spazio di probabilità $(\Omega,\mathcal F,P)$ (notebook 17.1). Per ogni $\omega\in\Omega$ fissato, la funzione $n\mapsto X_n(\omega)$ si chiama **traiettoria** (o realizzazione) del processo: un singolo "scenario" possibile, mentre il processo nel suo complesso descrive **tutti** gli scenari contemporaneamente, con le rispettive probabilità.

Le **distribuzioni finito-dimensionali** del processo sono le distribuzioni congiunte $\mu_{(X_{n_1},\ldots,X_{n_k})}$ per ogni scelta finita di indici $n_1<\cdots<n_k$ (nel senso del notebook 17.6, esteso a più di due variabili): determinano, in un senso preciso che esula dai dettagli tecnici di questo corso, l'intera "legge" del processo.

### La passeggiata aleatoria semplice

Siano $\xi_1,\xi_2,\ldots$ i.i.d. con $P(\xi_i=1)=P(\xi_i=-1)=1/2$ (variabili di Rademacher: una Bernoulli, notebook 17.3, riscalata a valori $\pm1$ invece di $0,1$). La **passeggiata aleatoria semplice** è
$$S_0=0,\qquad S_n=\xi_1+\cdots+\xi_n\quad(n\ge1).$$

$S_n$ descrive la posizione, dopo $n$ passi, di un punto che si sposta di $+1$ o $-1$ con probabilità uguale a ogni passo — il modello discreto più semplice di moto casuale.

**Proprietà elementari.** $E[\xi_i]=0$, $\mathrm{Var}(\xi_i)=E[\xi_i^2]-E[\xi_i]^2=1-0=1$ (essendo $\xi_i^2=1$ sempre). Per linearità (notebook 17.4) e indipendenza (notebook 17.6): $E[S_n]=0$, $\mathrm{Var}(S_n)=n\cdot1=n$ — la varianza cresce **linearmente** col numero di passi, quindi la dispersione tipica di $S_n$ (la deviazione standard $\sqrt n$) cresce come $\sqrt n$, non come $n$: la passeggiata aleatoria si allontana dall'origine molto più lentamente di un moto a velocità costante.

### La proprietà di Markov (informale)

**Osservazione.** $S_{n+1}=S_n+\xi_{n+1}$, dove $\xi_{n+1}$ è indipendente da $S_0,\ldots,S_n$ (essendo $\xi_{n+1}$ indipendente da $\xi_1,\ldots,\xi_n$, che determinano $S_0,\ldots,S_n$). Questo significa che, **dato il valore presente $S_n$**, il comportamento futuro della passeggiata non dipende in alcun modo da come si è arrivati a quel valore: è la **proprietà di Markov**, la caratteristica che definisce un'intera classe di processi stocastici (le catene di Markov), il cui studio sistematico va oltre l'introduzione di questo notebook.

### Il teorema centrale del limite applicato alla passeggiata aleatoria

Le ipotesi del notebook 17.9 sono soddisfatte con $\mu=0,\sigma^2=1$: quindi
$$\frac{S_n}{\sqrt n}=\frac{S_n-n\mu}{\sigma\sqrt n}\ \xrightarrow{d}\ \mathrm N(0,1).$$

**Perché questo è più di un'applicazione diretta del CLT.** Il riscalamento $S_n/\sqrt n$ non riguarda solo la distribuzione di $S_n$ a un singolo istante $n$ fissato: un risultato più profondo — il **teorema di Donsker** (o principio di invarianza, che enunciamo senza dimostrazione, essendo la dimostrazione rigorosa ben oltre lo scopo di questo corso) — afferma che l'**intera traiettoria riscalata** $t\mapsto S_{\lfloor nt\rfloor}/\sqrt n$ (interpolata linearmente, per $t\in[0,1]$), come processo a tempo continuo, converge (in un senso che generalizza la convergenza in distribuzione del notebook 17.7 a spazi di traiettorie) al **moto browniano standard** su $[0,1]$: un processo a tempo continuo con incrementi gaussiani indipendenti, traiettorie continue ma non derivabili in alcun punto, che è l'oggetto fondamentale del calcolo stocastico moderno (equazioni differenziali stocastiche, finanza matematica, fisica statistica).

### Uno sguardo indietro e uno in avanti

Questo notebook chiude il Modulo 17 collegando, in un'unica immagine, quasi tutto il percorso del corso: la passeggiata aleatoria è costruita da variabili i.i.d. elementari (Modulo 0-1: contare $+1,-1$), la sua varianza cresce grazie all'indipendenza (notebook 17.6), il suo comportamento asintotico è governato dal teorema centrale del limite (notebook 17.9, che a sua volta poggia sulla trasformata di Fourier del Modulo 15 e sull'analisi complessa del Modulo 9), e il suo limite continuo (il moto browniano) è un oggetto le cui traiettorie, pur continue, sfidano l'intuizione del calcolo differenziale del Modulo 5 — un fenomeno che motiva interi rami della matematica moderna, oltre lo scopo di questo corso introduttivo. Contare i cammini di una passeggiata aleatoria è, non a caso, anche il primo problema del prossimo modulo, dove il **Modulo 18 — Combinatoria e teoria dei numeri** fornirà gli strumenti per contarli esattamente, non solo asintoticamente.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simuliamo diverse traiettorie della passeggiata aleatoria semplice,
# e verifichiamo che Var(S_n) cresce linearmente in n.

np.random.seed(9)
n_passi = 2000
n_traiettorie = 8

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for _ in range(n_traiettorie):
    xi = np.random.choice([-1, 1], size=n_passi)
    S = np.concatenate([[0], np.cumsum(xi)])
    axes[0].plot(S, linewidth=0.8, alpha=0.7)
axes[0].set_title("Traiettorie della passeggiata aleatoria semplice")
axes[0].set_xlabel("n")
axes[0].set_ylabel("S_n")

# Verifica Var(S_n) = n
n_test = np.array([10, 50, 100, 500, 1000])
n_simulazioni = 20000
varianze_stimate = []
for n in n_test:
    xi = np.random.choice([-1, 1], size=(n_simulazioni, n))
    S_n = xi.sum(axis=1)
    varianze_stimate.append(np.var(S_n))

axes[1].plot(n_test, varianze_stimate, 'o-', label='Var(S_n) stimata')
axes[1].plot(n_test, n_test, 'r--', label='y = n (teorica)')
axes[1].set_xlabel("n")
axes[1].set_ylabel("Var(S_n)")
axes[1].legend()
axes[1].set_title("La varianza cresce linearmente in n")
plt.tight_layout()
plt.show()

for n, v in zip(n_test, varianze_stimate):
    print(f"n={n}: Var(S_n) stimata = {v:.2f}  (teorica: {n})")
```

## Esercizio svolto e commentato

**Problema:** usa il teorema centrale del limite per stimare $P(S_{100}>10)$ per la passeggiata aleatoria semplice, e confronta con una simulazione Monte Carlo.

**Ragionamento:**

**Passo 1 — richiamiamo i parametri.** $\mu=E[\xi_i]=0$, $\sigma^2=\mathrm{Var}(\xi_i)=1$, $n=100$.

**Passo 2 — standardizziamo.** $P(S_{100}>10)=P\Big(\dfrac{S_{100}}{\sqrt{100}}>\dfrac{10}{\sqrt{100}}\Big)=P\Big(\dfrac{S_{100}}{10}>1\Big)$.

**Passo 3 — applichiamo l'approssimazione del CLT (notebook 17.9).** Per $n=100$ (abbastanza grande), $S_{100}/\sqrt{100}=S_{100}/10\approx\mathrm N(0,1)$ in distribuzione. Quindi $P(S_{100}/10>1)\approx P(Z>1)=1-\Phi(1)\approx1-0.8413=0.1587$ (lo stesso valore numerico $\Phi(1)$ usato nel notebook 17.9, esercizio 3).

**Conclusione (da verificare nella cella seguente):** ci aspettiamo che la probabilità simulata di $S_{100}>10$ sia vicina a $0.1587$ — una buona approssimazione nonostante $\xi_i$ non sia affatto gaussiana (è a due soli valori, $\pm1$): esattamente la manifestazione, su questo esempio concreto, dell'universalità del teorema centrale del limite già osservata nel notebook 17.9 con la Bernoulli.

```python
import numpy as np
from scipy.stats import norm

n = 100
soglia = 10

# Approssimazione CLT
z = soglia / np.sqrt(n)
P_approssimata = 1 - norm.cdf(z)
print(f"P(S_100 > 10) approssimata (CLT): {P_approssimata:.4f}")

# Simulazione Monte Carlo diretta
np.random.seed(10)
n_simulazioni = 2_000_000
xi = np.random.choice([-1, 1], size=(n_simulazioni, n))
S_100 = xi.sum(axis=1)
P_simulata = np.mean(S_100 > soglia)
print(f"P(S_100 > 10) simulata (Monte Carlo): {P_simulata:.4f}")
```

3. Usa la proprietà di Markov (teoria) per spiegare intuitivamente perché $E[S_{n+1}\mid S_1,\ldots,S_n]=S_n$ (un processo con questa proprietà si chiama **martingala**): il valore atteso del passo successivo, dato tutto il passato, è esattamente il valore attuale, senza alcuna deriva sistematica.
4. *(più impegnativo)* Il numero di traiettorie della passeggiata aleatoria che vanno da $S_0=0$ a $S_n=k$ in esattamente $n$ passi è $\binom n{(n+k)/2}$ (se $n+k$ è pari, altrimenti $0$: servono $(n+k)/2$ passi $+1$ e $(n-k)/2$ passi $-1$). ## Riepilogo finale

- Un **processo stocastico** è una famiglia di variabili aleatorie indicizzate dal tempo, tutte definite sullo stesso spazio di probabilità; le sue **traiettorie** sono le realizzazioni per ogni singolo esito $\omega$.
- La **passeggiata aleatoria semplice** $S_n=\xi_1+\cdots+\xi_n$, costruita da incrementi i.i.d. $\pm1$, ha varianza $n$: si allontana dall'origine come $\sqrt n$, non come $n$.
- La **proprietà di Markov** (il futuro dipende dal passato solo attraverso il presente) rende la passeggiata aleatoria l'esempio prototipo di martingala e di catena di Markov.
- Il **teorema centrale del limite**, applicato a $S_n/\sqrt n$, converge alla normale standard; il teorema di Donsker (enunciato senza dimostrazione) estende questa convergenza all'intera traiettoria, anticipando il moto browniano come limite continuo.
- Contare esattamente le traiettorie con proprietà prescritte (come nell'esercizio 4) è un problema di calcolo combinatorio: il ponte naturale verso il prossimo modulo.
