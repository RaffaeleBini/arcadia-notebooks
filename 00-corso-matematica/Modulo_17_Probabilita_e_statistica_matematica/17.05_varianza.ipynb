# 17.5 — Varianza

**Prerequisiti:** 17.4 Valore atteso, 13.8 Spazi Lᵖ, 14.3 Spazi di Hilbert.


## Perché ci interessa

$E[X]$ riassume una variabile aleatoria con un solo numero, ma nasconde informazione: due variabili molto diverse (una quasi costante, una molto dispersa) possono avere lo stesso valore atteso. Serve una seconda quantità che misuri **quanto** $X$ si discosta, in media, dal proprio valore atteso: la **varianza**.

Il punto di vista più illuminante, qui, non è il calcolo algebrico ma quello **geometrico**: il notebook 14.3 ha mostrato che $L^2(X,\mu)$ è uno spazio di Hilbert, con prodotto scalare $\langle f,g\rangle=\int fg\,d\mu$. Applicato allo spazio di probabilità $(\Omega,\mathcal F,P)$, questo dice che $L^2(\Omega,P)$ — le variabili aleatorie con $E[X^2]<\infty$ — è esso stesso uno spazio di Hilbert, e la varianza altro non è che **il quadrato della distanza, in questo spazio, tra $X$ e la costante $E[X]$**.

## Teoria, passo dopo passo

### Definizione

Per $X\in L^2(\Omega,P)$ (cioè $E[X^2]<\infty$, notebook 13.8), la **varianza** è
$$\mathrm{Var}(X)=E\big[(X-E[X])^2\big].$$

**Formula alternativa.** $\mathrm{Var}(X)=E[X^2]-E[X]^2$.

**Dimostrazione.** Sviluppando il quadrato (linearità di $E$, notebook 17.4): $E[(X-E[X])^2]=E[X^2-2X\,E[X]+E[X]^2]=E[X^2]-2E[X]\,E[X]+E[X]^2=E[X^2]-2E[X]^2+E[X]^2=E[X^2]-E[X]^2$ (usando che $E[X]$ è una costante, quindi esce dal valore atteso per linearità: $E[2X\,E[X]]=2E[X]\cdot E[X]$). $\blacksquare$

La **deviazione standard** è $\sigma(X)=\sqrt{\mathrm{Var}(X)}$, nella stessa unità di misura di $X$ (a differenza della varianza, che è in unità al quadrato).

### Il punto di vista di Hilbert: varianza come distanza al quadrato

In $L^2(\Omega,P)$ (notebook 14.3, l'esempio fondamentale di spazio di Hilbert applicato qui a $\mu=P$), la norma è $\|X\|_2=\sqrt{E[X^2]}$ (notazione del notebook 13.8). La costante $E[X]$ è essa stessa un elemento di $L^2$ (una variabile aleatoria costante ha $E[E[X]^2]=E[X]^2<\infty$). Allora:
$$\mathrm{Var}(X)=E\big[(X-E[X])^2\big]=\|X-E[X]\|_2^2.$$

**Interpretazione.** La varianza è il quadrato della **distanza**, nello spazio di Hilbert $L^2(\Omega,P)$, tra la variabile aleatoria $X$ e la sua "migliore approssimazione costante" $E[X]$. Questo punto di vista spiega perché $E[X]$ (non, ad esempio, la mediana) è la costante "più vicina" a $X$ in questo senso: il notebook 14.3 (teorema della proiezione ortogonale) garantisce che la proiezione di $X$ sul sottospazio delle costanti è **l'unica** costante che minimizza la distanza $L^2$, e questa proiezione è esattamente $E[X]$ (una verifica diretta, lasciata come esercizio).

### Covarianza e Cauchy-Schwarz

Per $X,Y\in L^2(\Omega,P)$, la **covarianza** è $\mathrm{Cov}(X,Y)=E[XY]-E[X]E[Y]=E\big[(X-E[X])(Y-E[Y])\big]$ (stessa manipolazione algebrica della varianza) — esattamente il **prodotto scalare** $\langle X-E[X],Y-E[Y]\rangle$ in $L^2(\Omega,P)$ (notebook 14.3). In particolare $\mathrm{Cov}(X,X)=\mathrm{Var}(X)$.

**Teorema.** $|\mathrm{Cov}(X,Y)|\le\sigma(X)\sigma(Y)$.

**Dimostrazione.** Applicando la disuguaglianza di Cauchy-Schwarz (notebook 14.3) ai vettori $X-E[X]$ e $Y-E[Y]$ di $L^2(\Omega,P)$: $|\langle X-E[X],Y-E[Y]\rangle|\le\|X-E[X]\|_2\|Y-E[Y]\|_2$, cioè $|\mathrm{Cov}(X,Y)|\le\sqrt{\mathrm{Var}(X)}\sqrt{\mathrm{Var}(Y)}=\sigma(X)\sigma(Y)$. $\blacksquare$

Il **coefficiente di correlazione** $\rho(X,Y)=\dfrac{\mathrm{Cov}(X,Y)}{\sigma(X)\sigma(Y)}$ (per $\sigma(X),\sigma(Y)>0$) soddisfa quindi $|\rho(X,Y)|\le1$ — la stessa disuguaglianza soddisfatta dal coseno dell'angolo tra due vettori (notebook 6.20), non a caso: $\rho(X,Y)$ **è** il coseno dell'angolo tra $X-E[X]$ e $Y-E[Y]$ in $L^2(\Omega,P)$.

### Proprietà algebriche

- $\mathrm{Var}(aX+b)=a^2\mathrm{Var}(X)$ per costanti $a,b$ (una traslazione non cambia la dispersione, un riscalamento la cambia quadraticamente — verifica diretta dalla definizione).
- $\mathrm{Var}(X+Y)=\mathrm{Var}(X)+\mathrm{Var}(Y)+2\mathrm{Cov}(X,Y)$ (l'analogo probabilistico dell'espansione bilineare $\|u+v\|^2=\|u\|^2+\|v\|^2+2\langle u,v\rangle$ usata nella dimostrazione della legge del parallelogramma, notebook 14.3, applicata a $u=X-E[X],v=Y-E[Y]$).

```python
import numpy as np

# Verifica numerica: Var(X) come distanza L^2 al quadrato, e disuguaglianza
# di Cauchy-Schwarz per la covarianza, su un esempio simulato.

np.random.seed(1)
n = 2_000_000
X = np.random.exponential(scale=2.0, size=n)
Y = X + np.random.normal(scale=1.5, size=n)  # Y correlata con X

E_X, E_Y = np.mean(X), np.mean(Y)
Var_X = np.mean((X - E_X)**2)
Var_Y = np.mean((Y - E_Y)**2)
Cov_XY = np.mean((X - E_X) * (Y - E_Y))

print(f"E[X] = {E_X:.4f}, Var(X) = {Var_X:.4f}")
print(f"E[Y] = {E_Y:.4f}, Var(Y) = {Var_Y:.4f}")
print(f"Cov(X,Y) = {Cov_XY:.4f}")

sigma_X, sigma_Y = np.sqrt(Var_X), np.sqrt(Var_Y)
print(f"\n|Cov(X,Y)| = {abs(Cov_XY):.4f}")
print(f"sigma(X)*sigma(Y) = {sigma_X*sigma_Y:.4f}")
print(f"Cauchy-Schwarz verificata: {abs(Cov_XY) <= sigma_X*sigma_Y}")

rho = Cov_XY / (sigma_X * sigma_Y)
print(f"\nCoefficiente di correlazione rho = {rho:.4f}  (deve stare in [-1,1])")

# Verifica di Var(X+Y) = Var(X)+Var(Y)+2Cov(X,Y)
Var_somma_diretta = np.mean(((X+Y) - np.mean(X+Y))**2)
Var_somma_formula = Var_X + Var_Y + 2*Cov_XY
print(f"\nVar(X+Y) diretta: {Var_somma_diretta:.4f}")
print(f"Var(X)+Var(Y)+2Cov(X,Y): {Var_somma_formula:.4f}")
```

## Esercizio svolto e commentato

**Problema:** calcola $\mathrm{Var}(X)$ per $X\sim\mathrm U(a,b)$, usando la formula $E[X^2]-E[X]^2$.

**Ragionamento:**

**Passo 1 — richiamiamo $E[X]$.** Dal notebook 17.4: $E[X]=\dfrac{a+b}2$.

**Passo 2 — calcoliamo $E[X^2]$.** $E[X^2]=\displaystyle\int_a^bx^2\cdot\frac1{b-a}\,dx=\frac1{b-a}\Big[\frac{x^3}3\Big]_a^b=\frac{b^3-a^3}{3(b-a)}$.

**Passo 3 — semplifichiamo usando la fattorizzazione della differenza di cubi.** $b^3-a^3=(b-a)(b^2+ab+a^2)$ (identità algebrica, verificabile sviluppando il prodotto a destra), quindi $E[X^2]=\dfrac{(b-a)(a^2+ab+b^2)}{3(b-a)}=\dfrac{a^2+ab+b^2}3$.

**Passo 4 — calcoliamo $E[X]^2$.** $E[X]^2=\Big(\dfrac{a+b}2\Big)^2=\dfrac{a^2+2ab+b^2}4$.

**Passo 5 — sottraiamo, con denominatore comune $12$.** $\mathrm{Var}(X)=\dfrac{a^2+ab+b^2}3-\dfrac{a^2+2ab+b^2}4=\dfrac{4(a^2+ab+b^2)-3(a^2+2ab+b^2)}{12}=\dfrac{4a^2+4ab+4b^2-3a^2-6ab-3b^2}{12}=\dfrac{a^2-2ab+b^2}{12}=\dfrac{(a-b)^2}{12}$.

**Conclusione:** $\mathrm{Var}(X)=\dfrac{(b-a)^2}{12}$: la varianza cresce col **quadrato** della lunghezza dell'intervallo — raddoppiare l'ampiezza dell'intervallo quadruplica la varianza, coerente con la proprietà $\mathrm{Var}(aX)=a^2\mathrm{Var}(X)$ vista nella teoria (un intervallo $[a,b]$ più ampio è, a meno di traslazione, un riscalamento di uno più stretto).

```python
import sympy as sp

x, a, b = sp.symbols('x a b', real=True)

E_X = sp.Rational(1,1)*sp.integrate(x/(b-a), (x, a, b))
E_X2 = sp.integrate(x**2/(b-a), (x, a, b))

Var_X = sp.simplify(E_X2 - E_X**2)
print("E[X] =", sp.simplify(E_X))
print("E[X^2] =", sp.simplify(E_X2))
print("Var(X) = E[X^2] - E[X]^2 =", Var_X)
print("Confronto con (b-a)^2/12:", sp.simplify(Var_X - (b-a)**2/12) == 0)
```

## Esercizi da risolvere

1. Calcola $\mathrm{Var}(X)$ per $X\sim\mathrm{Exp}(\lambda)$, usando $E[X^2]=\displaystyle\int_0^\infty x^2\lambda e^{-\lambda x}\,dx$ (integrazione per parti due volte, come nell'esercizio svolto del notebook 17.4, applicata due volte).
2. Dimostra la proprietà $\mathrm{Var}(aX+b)=a^2\mathrm{Var}(X)$ direttamente dalla definizione $\mathrm{Var}(X)=E[(X-E[X])^2]$.
3. Se $X,Y$ sono tali che $Y=cX$ per una costante $c\ne0$ (perfettamente dipendenti, in modo lineare), verifica che $|\rho(X,Y)|=1$ — il caso di uguaglianza nella disuguaglianza di Cauchy-Schwarz, che nel notebook 14.3 corrisponde a vettori paralleli.
4. *(più impegnativo)* Dimostra la **disuguaglianza di Chebyshev**: per ogni $\varepsilon>0$, $P(|X-E[X]|\ge\varepsilon)\le\dfrac{\mathrm{Var}(X)}{\varepsilon^2}$. Suggerimento: applica la disuguaglianza di Markov (per $Z\ge0$, $P(Z\ge\varepsilon^2)\le E[Z]/\varepsilon^2$ — dimostrala prima, usando $Z\ge\varepsilon^2\mathbb 1_{\{Z\ge\varepsilon^2\}}$ e la monotonia di $E$) alla variabile $Z=(X-E[X])^2$.

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** $E[X^2]=\displaystyle\int_0^\infty x^2\lambda e^{-\lambda x}\,dx$. Per parti con $u=x^2,dv=\lambda e^{-\lambda x}dx$: $=\big[-x^2e^{-\lambda x}\big]_0^\infty+\int_0^\infty2xe^{-\lambda x}\,dx=0+\frac2\lambda\int_0^\infty x\lambda e^{-\lambda x}\,dx=\frac2\lambda\cdot\frac1\lambda=\frac2{\lambda^2}$ (riconoscendo, nell'ultimo integrale, esattamente $E[X]$ per l'esponenziale, notebook 17.4). Quindi $\mathrm{Var}(X)=E[X^2]-E[X]^2=\dfrac2{\lambda^2}-\dfrac1{\lambda^2}=\dfrac1{\lambda^2}$.

**2.** $E[aX+b]=aE[X]+b$ (linearità, notebook 17.4). $\mathrm{Var}(aX+b)=E\big[(aX+b-aE[X]-b)^2\big]=E\big[(aX-aE[X])^2\big]=E\big[a^2(X-E[X])^2\big]=a^2E\big[(X-E[X])^2\big]=a^2\mathrm{Var}(X)$ (linearità applicata alla costante $a^2$).

**3.** $\mathrm{Cov}(X,Y)=\mathrm{Cov}(X,cX)=E[XcX]-E[X]E[cX]=cE[X^2]-cE[X]^2=c\,\mathrm{Var}(X)$. $\sigma(Y)=\sqrt{\mathrm{Var}(cX)}=\sqrt{c^2\mathrm{Var}(X)}=|c|\sigma(X)$ (esercizio 2, con $b=0$). Quindi $\rho(X,Y)=\dfrac{c\,\mathrm{Var}(X)}{\sigma(X)\cdot|c|\sigma(X)}=\dfrac{c\,\mathrm{Var}(X)}{|c|\,\mathrm{Var}(X)}=\dfrac{c}{|c|}=\pm1$: $|\rho(X,Y)|=1$, coerentemente con l'interpretazione geometrica (essendo $Y-E[Y]=c(X-E[X])$, i due vettori sono paralleli in $L^2$, l'angolo tra loro è $0$ o $\pi$, quindi il coseno è $\pm1$).

**4.** Disuguaglianza di Markov: per $Z\ge0$ e $t>0$, $Z\ge t\,\mathbb 1_{\{Z\ge t\}}$ sempre (se $Z\ge t$, il membro destro è $t\le Z$; se $Z<t$, il membro destro è $0\le Z$). Applicando $E[\cdot]$ (monotonia, notebook 13.4): $E[Z]\ge t\,E[\mathbb 1_{\{Z\ge t\}}]=t\,P(Z\ge t)$, cioè $P(Z\ge t)\le E[Z]/t$. Applicando questo con $Z=(X-E[X])^2\ge0$ e $t=\varepsilon^2$: $P\big((X-E[X])^2\ge\varepsilon^2\big)\le\dfrac{E[(X-E[X])^2]}{\varepsilon^2}=\dfrac{\mathrm{Var}(X)}{\varepsilon^2}$. Ma $(X-E[X])^2\ge\varepsilon^2\iff|X-E[X]|\ge\varepsilon$ (radice quadrata, funzione crescente su $[0,\infty)$, notebook 3.9), quindi $P(|X-E[X]|\ge\varepsilon)\le\dfrac{\mathrm{Var}(X)}{\varepsilon^2}$: la disuguaglianza di Chebyshev.

## Riepilogo finale

- La **varianza** $\mathrm{Var}(X)=E[(X-E[X])^2]=E[X^2]-E[X]^2$ è, geometricamente, il quadrato della distanza $L^2$ (notebook 14.3) tra $X$ e la costante $E[X]$.
- La **covarianza** è il prodotto scalare $\langle X-E[X],Y-E[Y]\rangle$ in $L^2(\Omega,P)$; la disuguaglianza di Cauchy-Schwarz (notebook 14.3) dà immediatamente $|\mathrm{Cov}(X,Y)|\le\sigma(X)\sigma(Y)$.
- Il **coefficiente di correlazione** $\rho(X,Y)\in[-1,1]$ è, letteralmente, il coseno dell'angolo tra $X-E[X]$ e $Y-E[Y]$; $|\rho|=1$ esattamente quando $Y$ è funzione lineare di $X$.
- $\mathrm{Var}(X+Y)=\mathrm{Var}(X)+\mathrm{Var}(Y)+2\mathrm{Cov}(X,Y)$: l'analogo probabilistico della legge del parallelogramma.
- La **disuguaglianza di Chebyshev**, derivata dalla più elementare disuguaglianza di Markov, controlla quantitativamente quanto $X$ possa allontanarsi dalla propria media: sarà lo strumento chiave per la legge dei grandi numeri (notebook 17.8).
