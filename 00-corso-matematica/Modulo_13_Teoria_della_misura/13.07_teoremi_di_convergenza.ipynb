# 13.7 — Teoremi di convergenza

**Prerequisiti:** 13.4 Integrale di Lebesgue, 13.5 Convergenza quasi ovunque, 10.16 Convergenza uniforme.


## Perché ci interessa

Nel notebook 10.16 abbiamo dimostrato che la convergenza **uniforme** basta per scambiare limite e integrale (di Riemann): $\lim_n\int f_n=\int\lim_n f_n$. Ma la convergenza uniforme è una richiesta molto forte, spesso non soddisfatta nella pratica. Il vero potere dell'integrale di Lebesgue (notebook 13.4) emerge qui: con la sola convergenza **quasi ovunque** (notebook 13.5), molto più debole, si riescono comunque a scambiare limite e integrale — a patto di rispettare certe condizioni precise. Questo notebook presenta i tre teoremi cardine (convergenza monotona, lemma di Fatou, convergenza dominata) che rendono l'integrale di Lebesgue lo strumento standard in analisi, probabilità e equazioni alle derivate parziali.

## Teoria, passo dopo passo

### Teorema della convergenza monotona (Beppo Levi)

**Teorema.** Se $0\le f_1\le f_2\le\cdots$ (successione crescente di funzioni misurabili non negative) e $f_n\to f$ quasi ovunque, allora $\int f_n\,d\mu\to\int f\,d\mu$.

**Dimostrazione (schema).** Essendo $f_n\le f$ per ogni $n$ (limite di successione crescente), $\int f_n\,d\mu\le\int f\,d\mu$ per ogni $n$ (monotonia dell'integrale, conseguenza diretta della definizione tramite estremo superiore sulle funzioni semplici, notebook 13.4): dunque $\lim_n\int f_n\,d\mu\le\int f\,d\mu$ (il limite esiste in $[0,+\infty]$, essendo la successione $\int f_n\,d\mu$ anch'essa crescente per monotonia). Per la disuguaglianza opposta, si usa la costruzione delle funzioni semplici approssimanti (notebook 13.4): fissata una funzione semplice $s\le f$ e $c\in(0,1)$, si mostra che gli insiemi $E_n=\{x:f_n(x)\ge cs(x)\}$ sono crescenti con unione $X$ (essendo $f_n\to f\ge s>cs$ quasi ovunque), quindi $\int f_n\,d\mu\ge c\int_{E_n}s\,d\mu\to c\int s\,d\mu$ per la continuità dal basso della misura (notebook 13.2, applicata a ciascun pezzo della funzione semplice $s$). Facendo variare $s$ e $c\to1$, si ottiene $\lim_n\int f_n\,d\mu\ge\int f\,d\mu$. $\blacksquare$

**Perché serve la monotonia crescente.** Questo teorema è ciò che ha reso rigorosa, nel notebook 13.4, l'affermazione "$\int f\,d\mu=\lim_n\int s_n\,d\mu$" per la successione crescente di funzioni semplici $s_n\uparrow f$: la definizione stessa dell'integrale di Lebesgue per funzioni generiche **presuppone** questo teorema per essere ben posta (indipendente dalla particolare successione approssimante scelta).

### Il lemma di Fatou

**Lemma (Fatou).** Se $f_n\ge0$ sono misurabili, allora $\int\liminf_n f_n\,d\mu\le\liminf_n\int f_n\,d\mu$.

**Dimostrazione.** Poniamo $g_N=\inf_{n\ge N}f_n$ (misurabile, notebook 13.3). $(g_N)$ è **crescente** in $N$ (l'estremo inferiore su un insieme di indici più piccolo è maggiore o uguale), e $g_N\to\liminf_n f_n$ per definizione di liminf. Per il teorema della convergenza monotona: $\int g_N\,d\mu\to\int\liminf_n f_n\,d\mu$. Ma $g_N\le f_n$ per ogni $n\ge N$ (per definizione di estremo inferiore), quindi $\int g_N\,d\mu\le\int f_n\,d\mu$ per ogni $n\ge N$, cioè $\int g_N\,d\mu\le\inf_{n\ge N}\int f_n\,d\mu$. Passando al limite in $N$: $\int\liminf_n f_n\,d\mu=\lim_N\int g_N\,d\mu\le\lim_N\inf_{n\ge N}\int f_n\,d\mu=\liminf_n\int f_n\,d\mu$. $\blacksquare$

**Perché la disuguaglianza può essere stretta: un esempio.** Sulla successione "a macchina da scrivere" del notebook 13.6, $f_n\to0$ in misura ma $\liminf_n f_n(x)=0$ per ogni $x$ (essendo $f_n(x)=0$ infinite volte, per ogni $x$ fissato, come dimostrato nell'esercizio svolto del notebook 13.6). Dunque $\int\liminf_n f_n\,d\lambda=\int0\,d\lambda=0$. Ma $\int f_n\,d\lambda=$ larghezza del blocco $\to0$ anch'essa, quindi in questo esempio l'uguaglianza vale. Un esempio dove la disuguaglianza è **stretta**: $f_n=n\cdot\mathbb 1_{[0,1/n]}$ (notebook 13.5): $f_n\to0$ q.o. (quindi $\liminf_nf_n=0$ q.o.), $\int\liminf_nf_n\,d\lambda=0$; ma $\int f_n\,d\lambda=n\cdot(1/n)=1$ per ogni $n$, quindi $\liminf_n\int f_n\,d\lambda=1>0$: la "massa" dell'integrale sfugge al limite, concentrandosi su un insieme di misura sempre più piccola (un fenomeno chiamato "fuga di massa").

### Il teorema della convergenza dominata (Lebesgue)

**Teorema (convergenza dominata).** Se $f_n\to f$ quasi ovunque e esiste $g\ge0$ **integrabile** ($\int g\,d\mu<\infty$) con $|f_n|\le g$ quasi ovunque per ogni $n$, allora $\int f_n\,d\mu\to\int f\,d\mu$.

**Dimostrazione (usando Fatou due volte).** Applichiamo Fatou a $g+f_n\ge0$ (non negativa, essendo $f_n\ge-g$): $\int\liminf_n(g+f_n)\,d\mu\le\liminf_n\int(g+f_n)\,d\mu$, cioè $\int g\,d\mu+\int f\,d\mu\le\int g\,d\mu+\liminf_n\int f_n\,d\mu$ (usando $g+f_n\to g+f$ e la linearità, con $\int g\,d\mu$ finito quindi sottraibile), da cui $\int f\,d\mu\le\liminf_n\int f_n\,d\mu$. Applicando Fatou a $g-f_n\ge0$ (essendo $f_n\le g$): analogamente si ottiene $\int g\,d\mu-\int f\,d\mu\le\int g\,d\mu-\limsup_n\int f_n\,d\mu$, cioè $\limsup_n\int f_n\,d\mu\le\int f\,d\mu$. Combinando: $\limsup_n\int f_n\,d\mu\le\int f\,d\mu\le\liminf_n\int f_n\,d\mu$, che forza $\liminf_n=\limsup_n=\lim_n\int f_n\,d\mu=\int f\,d\mu$ (essendo sempre $\liminf\le\limsup$). $\blacksquare$

**Applicazione all'esempio "fuga di massa": la dominazione fallisce.** Per $f_n=n\cdot\mathbb 1_{[0,1/n]}$, non esiste alcuna $g$ integrabile con $f_n\le g$ per ogni $n$ (essendo $\sup_nf_n(x)=+\infty$ per $x$ vicino a $0$, dato che per ogni $x>0$ piccolo esiste $n$ con $1/n$ ancora più vicino a $0$ ma $f_n(x)$ arbitrariamente grande sulla porzione residua — più precisamente, la successione dei massimi $n$ non è dominabile da nessuna funzione fissata integrabile): questo spiega perché il teorema di convergenza dominata **non si applica** a questo esempio, coerentemente con il fatto che lì lo scambio di limite e integrale effettivamente fallisce.

```python
import numpy as np

# Verifichiamo numericamente il teorema di convergenza dominata su un esempio dove FUNZIONA
# f_n(x) = sin(nx)/n su [0, pi]: converge uniformemente (quindi anche q.o. e in misura) a 0
# dominata da g(x) = 1 (costante, integrabile su un dominio finito)

def f_n(x, n):
    return np.sin(n * x) / n

x = np.linspace(0, np.pi, 100000)
for n in [1, 5, 20, 100]:
    integrale = np.trapz(f_n(x, n), x)
    print(f"n={n}: integrale di f_n = {integrale:.6f}")

print(f"\nLimite atteso: integrale di f=0 su [0,pi] = 0")
print("|f_n(x)| = |sin(nx)|/n <= 1/n <= 1 = g(x), integrabile su [0,pi] (dominio finito)")
print("-> convergenza dominata si applica, e infatti l'integrale tende a 0")

# Confrontiamo con l'esempio "fuga di massa" dove la dominazione FALLISCE
def g_n(x, n):
    return np.where((x >= 0) & (x <= 1/n), n, 0)

print("\n--- Esempio senza dominazione (fuga di massa) ---")
for n in [1, 10, 100, 1000]:
    integrale = n * (1/n)  # esatto: n * larghezza del blocco
    print(f"n={n}: integrale di g_n = {integrale} (NON tende a 0, nonostante g_n->0 q.o.)")
```

## Esercizio svolto e commentato

**Problema:** usa il teorema della convergenza dominata per calcolare $\lim_{n\to\infty}\int_0^1\frac{n\sin(x/n)}{1+x^2}\,dx$.

**Ragionamento:**

**Passo 1 — identifichiamo $f_n$ e calcoliamo il limite puntuale.** $f_n(x)=\dfrac{n\sin(x/n)}{1+x^2}$. Per $x$ fissato in $[0,1]$, quando $n\to\infty$: $x/n\to0$, e usando il limite notevole $\sin(t)/t\to1$ per $t\to0$ (Modulo 4/5), scriviamo $n\sin(x/n)=x\cdot\dfrac{\sin(x/n)}{x/n}\to x\cdot1=x$ (per $x\ne0$; per $x=0$ banalmente $f_n(0)=0\to0=x$). Dunque $f_n(x)\to f(x)=\dfrac{x}{1+x^2}$ puntualmente su $[0,1]$ (quindi anche quasi ovunque).

**Passo 2 — troviamo una dominante integrabile.** Usiamo $|\sin(t)|\le|t|$ per ogni $t$ (disuguaglianza elementare, dimostrabile con gli strumenti del Modulo 5/10): $|n\sin(x/n)|\le n\cdot|x/n|=|x|=x$ (per $x\in[0,1]$). Dunque $|f_n(x)|\le\dfrac{x}{1+x^2}\le\dfrac{1}{1+x^2}\cdot1\le1=:g(x)$ per ogni $x\in[0,1]$ (usando $x\le1$ su questo dominio). $g(x)=1$ è integrabile su $[0,1]$ (dominio di misura finita, $\int_0^1 1\,dx=1<\infty$).

**Passo 3 — applichiamo il teorema.** Essendo $f_n\to f$ q.o. e $|f_n|\le g$ integrabile, per il teorema della convergenza dominata (teoria): $\lim_n\int_0^1 f_n\,dx=\int_0^1 f\,dx=\int_0^1\dfrac{x}{1+x^2}\,dx$.

**Passo 4 — calcoliamo l'integrale limite (elementare, sostituzione $u=1+x^2$, $du=2x\,dx$).** $\int_0^1\dfrac{x}{1+x^2}\,dx=\dfrac12\big[\ln(1+x^2)\big]_0^1=\dfrac12\ln2$.

**Conclusione:** $\lim_{n\to\infty}\int_0^1\dfrac{n\sin(x/n)}{1+x^2}\,dx=\dfrac{\ln2}{2}$. **Punto metodologico:** senza il teorema di convergenza dominata, calcolare direttamente questo limite (scambiando limite e integrale "a occhio") non sarebbe giustificato — il teorema fornisce esattamente la condizione (dominazione da $g$ integrabile) che rende lecito lo scambio, trasformando un limite di integrali complicati in un integrale di un limite semplice.

```python
import numpy as np
from scipy import integrate

def f_n(x, n):
    return n * np.sin(x / n) / (1 + x**2)

for n in [1, 5, 20, 100, 1000]:
    valore, _ = integrate.quad(f_n, 0, 1, args=(n,))
    print(f"n={n}: integrale = {valore:.6f}")

limite_teorico = 0.5 * np.log(2)
print(f"\nLimite teorico (ln(2)/2) = {limite_teorico:.6f}")
```

4. *(più impegnativo)* Dimostra che l'ipotesi di dominazione nel teorema di convergenza dominata è **necessaria**, usando l'esempio $f_n=n\cdot\mathbb 1_{[0,1/n]}$ della teoria: verifica esplicitamente che $\int f_n\,d\lambda\not\to\int f\,d\lambda$ (dove $f=0$ è il limite q.o.), e spiega perché questo non contraddice il lemma di Fatou (che vale sempre, dominazione o meno).

**1.** Per $x\in(0,1)$ fissato: $x^{1/n}=e^{(\ln x)/n}$, e $(\ln x)/n\to0$ quando $n\to\infty$ (essendo $\ln x$ un numero fissato, finito), quindi $x^{1/n}\to e^0=1$. Per $x=0$: $0^{1/n}=0$ per ogni $n$, quindi il limite è $0$ (ma questo è un singolo punto, di misura nulla). La successione è crescente in $n$ per $x\in(0,1)$ fissato (essendo $\ln x<0$, dividere per $n$ crescente rende $(\ln x)/n$ meno negativo, quindi $x^{1/n}$ cresce verso $1$). Per il teorema della convergenza monotona: $\lim_n\int_0^1x^{1/n}\,dx=\int_0^1 1\,dx=1$ (essendo il limite puntuale $1$ quasi ovunque, tranne in $x=0$).

**2.** Per $x$ fissato, $f_n(x)=1$ solo se $n\le x\le n+1$, cioè per **al più un** valore di $n$ (essendo gli intervalli $[n,n+1]$ disgiunti per $n$ interi distinti); per tutti gli altri $n$, $f_n(x)=0$. Dunque $\liminf_nf_n(x)=0$ per ogni $x$ (essendo $f_n(x)=0$ per infiniti $n$). $\int\liminf_nf_n\,d\lambda=\int0\,d\lambda=0$. D'altra parte, $\int f_n\,d\lambda=\lambda([n,n+1])=1$ per ogni $n$, quindi $\liminf_n\int f_n\,d\lambda=1$. Verifica di Fatou: $0=\int\liminf_nf_n\,d\lambda\le\liminf_n\int f_n\,d\lambda=1$ ✓ (disuguaglianza stretta, un altro esempio di "fuga di massa", qui verso $+\infty$ invece che verso un punto).

**3.** $f_n(x)=(1+x/n)^{-n}\to e^{-x}$ puntualmente per $x\ge0$ (limite notevole standard). Per la disuguaglianza di Bernoulli, $(1+x/n)^n\ge1+n\cdot(x/n)=1+x$ per $x\ge0$ (con uguaglianza solo per $n=1$), quindi $f_n(x)=(1+x/n)^{-n}\le(1+x)^{-1}$. Ma $(1+x)^{-1}$ **non** è integrabile su $[0,\infty)$ (il suo integrale diverge, essendo collegato alla serie armonica, notebook 4.9): serve una dominante migliore. Usando invece $(1+x/n)^n\ge e^{x/2}$ per $n$ abbastanza grande e $x$ in un range fissato (stima più fine, tecnica standard ma che omettiamo nel dettaglio), si trova una dominante integrabile $g(x)=e^{-x/2}$, e il teorema dà $\lim_n\int_0^\infty f_n\,dx=\int_0^\infty e^{-x}\,dx=1$.

**4.** $\int f_n\,d\lambda=n\cdot(1/n)=1$ per ogni $n$, quindi $\lim_n\int f_n\,d\lambda=1$. Ma $f_n\to0$ quasi ovunque (notebook 13.5), quindi $\int f\,d\lambda=\int0\,d\lambda=0$. Dunque $1=\lim_n\int f_n\,d\lambda\ne\int f\,d\lambda=0$: lo scambio di limite e integrale **fallisce**. Questo non contraddice Fatou, perché Fatou dà solo una **disuguaglianza** ($\int\liminf_nf_n\,d\lambda\le\liminf_n\int f_n\,d\lambda$, cioè $0\le1$, vera), non un'uguaglianza: Fatou vale **sempre** (per funzioni non negative, senza bisogno di dominazione), mentre il teorema di convergenza dominata richiede la dominazione **in più** per garantire l'**uguaglianza** nel limite, ipotesi che qui manca (nessuna $g$ integrabile domina $f_n=n\cdot\mathbb1_{[0,1/n]}$ per ogni $n$, come discusso nella teoria).

## Riepilogo finale

- Il **teorema della convergenza monotona** garantisce lo scambio di limite e integrale per successioni crescenti non negative: è il fondamento stesso della definizione dell'integrale di Lebesgue.
- Il **lemma di Fatou** dà sempre una disuguaglianza ($\int\liminf\le\liminf\int$) per funzioni non negative, senza altre ipotesi: la disuguaglianza può essere stretta quando c'è "fuga di massa".
- Il **teorema della convergenza dominata** (Lebesgue) garantisce l'uguaglianza $\lim\int f_n=\int\lim f_n$ sotto la sola ipotesi di convergenza quasi ovunque, purché esista una dominante integrabile fissata: la condizione che esclude la fuga di massa.
- Questi tre teoremi, insieme, sono lo strumento standard per giustificare rigorosamente lo scambio di limiti e integrali in tutta l'analisi moderna, molto più flessibili della sola convergenza uniforme (notebook 10.16).
