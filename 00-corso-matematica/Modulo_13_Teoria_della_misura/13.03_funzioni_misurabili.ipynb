# 13.3 — Funzioni misurabili

**Prerequisiti:** 13.2 Misure, 12.4 Continuità (topologica), 12.1 Spazi topologici.


## Perché ci interessa

Nel notebook 12.4 abbiamo definito la continuità come "la controimmagine di ogni aperto è aperta". Per costruire un integrale rispetto a una misura (il prossimo passo, notebook 13.4), serve una condizione analoga ma più debole: non ci serve che $f$ rispetti la topologia, ci basta che rispetti la **struttura di misurabilità** — cioè che le controimmagini degli insiemi "semplici" da misurare siano a loro volta misurabili. Questa nozione, la **misurabilità**, è più permissiva della continuità (funzioni discontinue, come quella di Dirichlet, possono essere misurabili) e chiude perfettamente il cerchio: costruiremo l'integrale di Lebesgue esattamente sulle funzioni misurabili, così come l'integrale di Riemann era costruito sulle funzioni (quasi ovunque) continue.

## Teoria, passo dopo passo

### Definizione di funzione misurabile

Siano $(X,\mathcal F)$ e $(Y,\mathcal G)$ spazi misurabili (notebook 13.1). Una funzione $f:X\to Y$ è **misurabile** se $f^{-1}(B)\in\mathcal F$ per ogni $B\in\mathcal G$. Nel caso più comune, $Y=\mathbb R$ con $\mathcal G=\mathcal B(\mathbb R)$ (Borel, notebook 13.1): $f:X\to\mathbb R$ è misurabile se $f^{-1}(B)\in\mathcal F$ per ogni boreliano $B$.

**Confronto diretto con la continuità topologica (notebook 12.4):** la definizione è identica nella forma — "la controimmagine di ogni insieme buono è buona" — ma qui "buono" significa "misurabile" (elemento della σ-algebra), non "aperto". Questo parallelismo non è casuale: **ogni funzione continua è misurabile**, come dimostriamo subito.

### Criterio pratico: basta controllare le semirette

**Teorema.** $f:X\to\mathbb R$ è misurabile se e solo se $f^{-1}((-\infty,a))\in\mathcal F$ per ogni $a\in\mathbb R$.

**Dimostrazione.** *(⇒)* Ogni semiretta $(-\infty,a)$ è aperta, quindi boreliana (notebook 13.1); dunque la sua controimmagine è misurabile, se $f$ lo è. *(⇐)* Supponiamo $f^{-1}((-\infty,a))\in\mathcal F$ per ogni $a$. La famiglia $\mathcal G=\{B\subseteq\mathbb R:f^{-1}(B)\in\mathcal F\}$ è una σ-algebra (verifica diretta: $f^{-1}(\mathbb R)=X\in\mathcal F$; $f^{-1}(\mathbb R\setminus B)=X\setminus f^{-1}(B)$, quindi chiusura per complementare eredita da $\mathcal F$; $f^{-1}(\bigcup_n B_n)=\bigcup_n f^{-1}(B_n)$, quindi chiusura per unioni numerabili eredita da $\mathcal F$ — le stesse identità insiemistiche usate nel notebook 12.4 per la continuità). Per ipotesi, $\mathcal G$ contiene tutte le semirette $(-\infty,a)$; essendo $\mathcal G$ una σ-algebra che le contiene, contiene anche $\sigma(\{(-\infty,a):a\in\mathbb R\})=\mathcal B(\mathbb R)$ (le semirette generano i boreliani: ogni intervallo aperto $(a,b)=(-\infty,b)\setminus(-\infty,a]$ si ottiene da esse, e gli aperti sono unioni numerabili di intervalli, notebook 12.2). Dunque $\mathcal G\supseteq\mathcal B(\mathbb R)$, cioè $f^{-1}(B)\in\mathcal F$ per ogni boreliano $B$: $f$ è misurabile. $\blacksquare$ Questo teorema è lo strumento pratico principale: per verificare la misurabilità, basta controllare un insieme molto più piccolo (le semirette) invece di tutti i boreliani.

### Ogni funzione continua è misurabile (rispetto a Borel)

**Corollario.** Se $X$ ha una topologia e $\mathcal F=\mathcal B(X)$ (Borel su $X$), ogni funzione continua $f:X\to\mathbb R$ è misurabile.

**Dimostrazione.** Per continuità (notebook 12.4), $f^{-1}((-\infty,a))$ è aperto per ogni $a$ (essendo $(-\infty,a)$ aperto in $\mathbb R$), quindi boreliano. Per il teorema, $f$ è misurabile. $\blacksquare$

### La funzione di Dirichlet è misurabile (ma non continua)

Sia $D:\mathbb R\to\mathbb R$, $D(x)=1$ se $x\in\mathbb Q$, $D(x)=0$ altrimenti. $D$ non è continua in **nessun** punto: per la densità di $\mathbb Q$ in $\mathbb R$ (notebook 10.9) e la densità del suo complementare (dimostrabile con lo stesso argomento, intercalando un irrazionale tra due razionali qualsiasi), ogni intorno di ogni punto contiene sia razionali sia irrazionali, quindi $D$ oscilla ovunque tra $0$ e $1$. Ma $D$ **è** misurabile: $D^{-1}((-\infty,a))=\emptyset$ se $a\le0$, $=\mathbb R\setminus\mathbb Q$ se $0<a\le1$, $=\mathbb R$ se $a>1$. Essendo $\mathbb R\setminus\mathbb Q$ boreliano (complementare di $\mathbb Q$, boreliano per l'esercizio 4 del notebook 13.1), tutte e tre le controimmagini sono in $\mathcal B(\mathbb R)$: per il teorema, $D$ è misurabile. **Questo esempio anticipa** il punto centrale del notebook 13.4: la teoria della misura riesce a "integrare" funzioni troppo discontinue per Riemann, proprio grazie a questa nozione più permissiva di misurabilità.

### Operazioni su funzioni misurabili

**Teorema (enunciato, dimostrazione per esercizio).** Se $f,g:X\to\mathbb R$ sono misurabili, allora $f+g$, $f\cdot g$, $\max(f,g)$, $\min(f,g)$ sono misurabili; e se $(f_n)$ è una successione di funzioni misurabili, allora $\sup_n f_n$, $\inf_n f_n$, $\limsup_n f_n$, $\liminf_n f_n$ sono misurabili (dove definite in $[-\infty,+\infty]$). Questa chiusura per limiti puntuali — anche infiniti, anche senza convergenza — è una differenza fondamentale rispetto alle funzioni continue: il limite puntuale di funzioni continue **non** è necessariamente continuo (esempio classico: $f_n(x)=x^n$ su $[0,1]$ converge puntualmente alla funzione discontinua che vale $0$ su $[0,1)$ e $1$ in $x=1$), ma il limite puntuale di funzioni misurabili **è sempre** misurabile.

```python
import numpy as np
import matplotlib.pyplot as plt

# Illustriamo che il limite puntuale di funzioni continue puo' non essere continuo,
# ma resta comunque misurabile (a differenza della continuita', che si perde nel limite)
x = np.linspace(0, 1, 500)

fig, ax = plt.subplots(figsize=(7, 5))
for n in [1, 2, 5, 20, 100]:
    ax.plot(x, x**n, label=f"f_{n}(x) = x^{n}", alpha=0.7)

ax.axhline(0, color='gray', linewidth=0.5)
ax.set_title("f_n(x)=x^n: ciascuna continua, ma il limite puntuale NON lo e'")
ax.legend()
plt.tight_layout()
plt.show()

def limite_puntuale(x):
    return 1.0 if x == 1 else 0.0

print("Limite puntuale: f(x)=0 per x in [0,1), f(1)=1")
print("Questa funzione e' discontinua in x=1, ma resta misurabile (Borel):")
print("f^-1((-inf, 0.5)) = [0,1), un intervallo, quindi boreliano")
```

## Esercizio svolto e commentato

**Problema:** dimostra che se $f:X\to\mathbb R$ è misurabile, allora anche $|f|$ e $f^2$ sono misurabili.

**Ragionamento:**

**Passo 1 — misurabilità di $|f|$: usiamo il criterio delle semirette.** Per il teorema della teoria, basta mostrare che $\{x:|f(x)|<a\}\in\mathcal F$ per ogni $a\in\mathbb R$. Se $a\le0$: $\{x:|f(x)|<a\}=\emptyset\in\mathcal F$ (assioma di σ-algebra, notebook 13.1). Se $a>0$: $|f(x)|<a\iff-a<f(x)<a$, quindi $\{x:|f(x)|<a\}=\{x:f(x)<a\}\cap\{x:f(x)>-a\}=f^{-1}((-\infty,a))\cap\big(X\setminus f^{-1}((-\infty,-a])\big)$.

**Passo 2 — verifichiamo che ogni pezzo è misurabile.** $f^{-1}((-\infty,a))\in\mathcal F$ per il criterio delle semirette (essendo $f$ misurabile). $f^{-1}((-\infty,-a])=\bigcap_n f^{-1}((-\infty,-a+1/n))$ (intersezione numerabile di semirette aperte che si restringono a quella chiusa — verifica diretta: $x\in$ intersezione $\iff f(x)<-a+1/n$ per ogni $n$ $\iff f(x)\le-a$), quindi misurabile (intersezione numerabile di misurabili, dedotta da De Morgan nel notebook 13.1). Il suo complementare è quindi misurabile (assioma 2 di σ-algebra).

**Passo 3 — combiniamo.** L'intersezione di due insiemi misurabili è misurabile (De Morgan, notebook 13.1): $\{x:|f(x)|<a\}\in\mathcal F$ per ogni $a$. Per il criterio delle semirette, $|f|$ è misurabile.

**Passo 4 — misurabilità di $f^2$.** $\{x:f(x)^2<a\}=\emptyset$ se $a\le0$; se $a>0$, $f(x)^2<a\iff|f(x)|<\sqrt a$, quindi $\{x:f(x)^2<a\}=\{x:|f(x)|<\sqrt a\}\in\mathcal F$ per il passo 3 (applicato con $\sqrt a$ al posto di $a$). Per il criterio delle semirette, $f^2$ è misurabile.

**Conclusione:** sia $|f|$ sia $f^2$ sono misurabili. **Punto metodologico:** questo esercizio mostra la tecnica standard per dimostrare la misurabilità di funzioni costruite componendo altre funzioni misurabili — ridurre sempre al criterio delle semirette, ed esprimere l'insieme richiesto tramite operazioni insiemistiche (unione, intersezione, complementare, numerabili) applicate a insiemi già noti come misurabili.

```python
import numpy as np

# Verifica numerica: la funzione |f| e f^2 hanno le controimmagini attese quando f e' misurabile
def f(x):
    return x - 2  # funzione continua (quindi misurabile), per semplicita'

# Controimmagine di (-inf, a) per |f|
a = 3
xs = np.linspace(-10, 10, 100000)
insieme_diretto = xs[np.abs(f(xs)) < a]
print(f"Range di {{x : |f(x)| < {a}}}: [{insieme_diretto.min():.3f}, {insieme_diretto.max():.3f}]")
print(f"Atteso: -1 < f(x) < 1 ancora, cioe' -a<f(x)<a -> {2-a} < x < {2+a}")

# Verifica per f^2
insieme_quadrato = xs[f(xs)**2 < a]
print(f"\nRange di {{x : f(x)^2 < {a}}}: [{insieme_quadrato.min():.3f}, {insieme_quadrato.max():.3f}]")
print(f"Atteso: |f(x)| < sqrt({a})={np.sqrt(a):.3f} -> {2-np.sqrt(a):.3f} < x < {2+np.sqrt(a):.3f}")
```

## Riepilogo finale

- Una funzione $f$ è **misurabile** se la controimmagine di ogni insieme misurabile del codominio è misurabile nel dominio — stessa forma della continuità (notebook 12.4), ma rispetto a σ-algebre invece che topologie.
- Il **criterio delle semirette** ($f^{-1}((-\infty,a))$ misurabile per ogni $a$) basta a garantire la misurabilità su tutto $\mathcal B(\mathbb R)$: uno strumento pratico decisivo.
- Ogni funzione continua è misurabile, ma la misurabilità è molto più permissiva: la funzione di Dirichlet, discontinua ovunque, è misurabile.
- Somme, prodotti, massimi, minimi, e soprattutto **limiti puntuali** (anche non uniformi) di funzioni misurabili restano misurabili — una chiusura molto più robusta di quella della continuità, ed è precisamente ciò che rende l'integrale di Lebesgue (notebook 13.4) più potente di quello di Riemann.
