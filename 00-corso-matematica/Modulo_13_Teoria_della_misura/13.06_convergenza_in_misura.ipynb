# 13.6 — Convergenza in misura

**Prerequisiti:** 13.5 Convergenza quasi ovunque, 13.2 Misure.


## Perché ci interessa

Nel notebook 13.5 abbiamo visto che la convergenza quasi ovunque è più debole di quella uniforme, ma resta comunque una nozione "puntuale": guarda cosa succede in **ogni singolo** $x$ (a parte un insieme trascurabile). C'è un modo completamente diverso di misurare la "vicinanza" tra $f_n$ e $f$: invece di controllare ogni punto, controlliamo **quanto è grande** (in misura) l'insieme dove $f_n$ si discosta significativamente da $f$. Questa è la **convergenza in misura**, e sorprendentemente **non implica né è implicata** dalla convergenza quasi ovunque in generale — le due nozioni catturano aspetti diversi del comportamento di una successione, e il controesempio della "macchina da scrivere" mostra con precisione dove divergono.

## Teoria, passo dopo passo

### Definizione di convergenza in misura

Una successione di funzioni misurabili $(f_n)$ **converge in misura** a $f$ se, per ogni $\varepsilon>0$: $\mu(\{x:|f_n(x)-f(x)|\ge\varepsilon\})\to0$ quando $n\to\infty$. Intuitivamente: la "quantità di spazio" dove $f_n$ differisce sensibilmente da $f$ diventa arbitrariamente piccola, anche se **quali** punti compongono quell'insieme può continuare a cambiare a ogni $n$ (a differenza della convergenza q.o., dove ogni singolo punto, fissato, deve stabilizzarsi).

### Convergenza q.o. su misura finita implica convergenza in misura

**Teorema.** Se $\mu(X)<\infty$ e $f_n\to f$ quasi ovunque, allora $f_n\to f$ in misura.

**Dimostrazione (schema, usando Egorov).** Fissato $\varepsilon,\delta>0$, per il teorema di Egorov (notebook 13.5) esiste $E$ con $\mu(X\setminus E)<\delta$ su cui $f_n\to f$ uniformemente; per $n$ abbastanza grande, $\sup_{x\in E}|f_n(x)-f(x)|<\varepsilon$, quindi $\{x:|f_n(x)-f(x)|\ge\varepsilon\}\subseteq X\setminus E$, dando $\mu(\{x:|f_n(x)-f(x)|\ge\varepsilon\})\le\mu(X\setminus E)<\delta$. Essendo $\delta$ arbitrario, il limite è $0$. $\blacksquare$ Questo teorema richiede **essenzialmente** $\mu(X)<\infty$ (come Egorov): il prossimo esempio mostra cosa succede quando manca la condizione di convergenza q.o. stessa, indipendentemente dalla finitezza.

### Il controesempio della "macchina da scrivere" (typewriter sequence)

Su $[0,1]$ con la misura di Lebesgue, costruiamo $(f_n)$ così: $f_1=\mathbb 1_{[0,1]}$; $f_2=\mathbb 1_{[0,1/2]}$, $f_3=\mathbb 1_{[1/2,1]}$; $f_4=\mathbb 1_{[0,1/4]}$, $f_5=\mathbb 1_{[1/4,1/2]}$, $f_6=\mathbb 1_{[1/2,3/4]}$, $f_7=\mathbb 1_{[3/4,1]}$; e così via, con blocchi di larghezza $1/2^k$ che "scorrono" attraverso $[0,1]$ come la testina di una macchina da scrivere, dimezzando la larghezza a ogni ciclo completo.

**Convergenza in misura a $0$.** Il blocco usato per $f_n$ ha larghezza $w_n\to0$ (al crescere di $n$, i cicli usano blocchi sempre più stretti): $\mu(\{x:|f_n(x)-0|\ge\varepsilon\})=\mu(\{x:f_n(x)=1\})=w_n\to0$ per ogni $\varepsilon\in(0,1)$ (essendo $f_n$ a valori $0$ o $1$). Dunque $f_n\to0$ **in misura**.

**Nessuna convergenza puntuale, in nessun punto.** Fissato **qualunque** $x\in[0,1]$, ogni ciclo (con blocchi di una data larghezza $1/2^k$) contiene esattamente un blocco che copre $x$: dunque $f_n(x)=1$ per infiniti $n$ (uno per ogni ciclo) e $f_n(x)=0$ per infiniti altri $n$. La successione numerica $f_n(x)$ oscilla per sempre tra $0$ e $1$, **senza mai convergere**, per **ogni** $x$ fissato. In particolare, $f_n$ **non converge quasi ovunque** a $0$ (l'insieme dove non converge è **tutto** $[0,1]$, non di misura nulla) — anzi, non converge puntualmente in nessun punto.

**Conclusione concettuale.** Questo esempio dimostra che convergenza in misura **non implica** convergenza quasi ovunque: le due nozioni sono genuinamente diverse. Il teorema precedente ci dice che c'è comunque un legame parziale — vale il viceversa (q.o. $\Rightarrow$ in misura, su misura finita) — ma non un'equivalenza.

### Un fatto di consolazione: sottosuccessioni convergenti q.o.

**Teorema (enunciato, dimostrazione fuori portata).** Se $f_n\to f$ in misura, esiste una **sottosuccessione** $(f_{n_k})$ che converge a $f$ **quasi ovunque**. Applicato alla macchina da scrivere: pur non convergendo q.o. l'intera successione, si può sempre estrarre una sottosuccessione (per esempio, prendendo un solo blocco per ciclo, sempre più piccolo e posizionato in modo che ogni punto sia coperto solo un numero finito di volte) che converge q.o. a $0$ — un fenomeno che ricorda da vicino Bolzano-Weierstrass (notebook 10.10): la successione intera "non si comporta bene", ma una sua sottosuccessione sì.

```python
import numpy as np
import matplotlib.pyplot as plt

def macchina_da_scrivere(n_max):
    # genera i blocchi [inizio, fine) della successione "a macchina da scrivere"
    blocchi = []
    k = 0
    while len(blocchi) < n_max:
        n_blocchi_ciclo = 2**k
        larghezza = 1 / n_blocchi_ciclo
        for i in range(n_blocchi_ciclo):
            if len(blocchi) >= n_max:
                break
            blocchi.append((i * larghezza, (i + 1) * larghezza, larghezza))
        k += 1
    return blocchi

blocchi = macchina_da_scrivere(15)
for n, (a, b, w) in enumerate(blocchi, 1):
    print(f"f_{n}: blocco [{a:.4f}, {b:.4f}), larghezza (= misura dove f_n=1) = {w:.4f}")

# Verifica: la larghezza (misura del sostegno) tende a 0
larghezze = [w for _, _, w in blocchi]
print(f"\nLe larghezze tendono a 0: {larghezze[-5:]}")

# Verifica: per un punto fissato x=0.3, f_n(0.3) oscilla per sempre tra 0 e 1
x_fissato = 0.3
valori_in_x = [1 if a <= x_fissato < b else 0 for a, b, w in blocchi]
print(f"\nValori di f_n({x_fissato}) per n=1..15: {valori_in_x}")
print("Non converge: continua a valere sia 0 sia 1 anche per n grandi")
```

## Esercizio svolto e commentato

**Problema:** verifica esplicitamente, calcolando i primi termini, che per $x=0.3$ la successione a macchina da scrivere $f_n(0.3)$ vale $1$ infinite volte e $0$ infinite volte, confermando che non converge in quel punto (né in nessun altro, per lo stesso argomento).

**Ragionamento:**

**Passo 1 — troviamo, per ogni "ciclo" $k$, quale blocco copre $x=0.3$.** Nel ciclo $k$ (blocchi di larghezza $1/2^k$), il blocco che copre $x=0.3$ è quello con indice $i=\lfloor 0.3\cdot2^k\rfloor$ (la parte intera di $x$ diviso la larghezza del blocco): per $k=0$: $i=0$ (blocco $[0,1)$, copre tutto); per $k=1$: $i=\lfloor0.6\rfloor=0$ (blocco $[0,0.5)$); per $k=2$: $i=\lfloor1.2\rfloor=1$ (blocco $[0.25,0.5)$); per $k=3$: $i=\lfloor2.4\rfloor=2$ (blocco $[0.25,0.375)$).

**Passo 2 — per ogni ciclo, esiste esattamente un $n$ (l'indice cumulativo) per cui $f_n(0.3)=1$.** Poiché ogni ciclo $k$ partiziona $[0,1)$ in $2^k$ blocchi disgiunti che coprono **tutto** l'intervallo, $x=0.3$ appartiene a **esattamente uno** di essi: dunque, per ogni $k=0,1,2,\ldots$, esiste esattamente un indice $n$ (quello del blocco del ciclo $k$ che copre $0.3$) con $f_n(0.3)=1$.

**Passo 3 — concludiamo che $f_n(0.3)=1$ per infiniti $n$.** Essendo i cicli infiniti ($k=0,1,2,\ldots$ senza fine), e ogni ciclo contribuendo (per il passo 2) esattamente un indice $n$ con $f_n(0.3)=1$, esistono **infiniti** indici $n$ con $f_n(0.3)=1$. Per tutti gli **altri** indici $n$ (gli altri blocchi dello stesso ciclo, che non coprono $0.3$), $f_n(0.3)=0$: essendo ogni ciclo composto da $2^k-1$ blocchi che non coprono $0.3$ contro $1$ che lo copre, anche gli indici con $f_n(0.3)=0$ sono infiniti.

**Conclusione:** $f_n(0.3)$ assume il valore $1$ infinite volte e il valore $0$ infinite volte: la successione numerica $(f_n(0.3))_n$ **non converge** (una successione convergente non può assumere due valori diversi infinite volte, per l'unicità del limite, Modulo 10). Essendo $x=0.3$ arbitrario nell'argomento (nessuna proprietà specifica di $0.3$ è stata usata, se non che sta in $[0,1)$), lo stesso vale per **ogni** $x\in[0,1)$: nessun punto converge.

```python
import numpy as np

def indice_blocco_copertura(x, k):
    n_blocchi = 2**k
    return int(x * n_blocchi)  # indice del blocco che copre x nel ciclo k

x = 0.3
print(f"x = {x}")
for k in range(6):
    i = indice_blocco_copertura(x, k)
    larghezza = 1 / 2**k
    print(f"Ciclo k={k}: blocco indice i={i}, intervallo [{i*larghezza:.4f}, {(i+1)*larghezza:.4f}), larghezza={larghezza:.4f}")

print(f"\nPer ogni ciclo k, x={x} e' coperto da esattamente 1 blocco su {2**5}=32 (nel ciclo k=5):")
print("-> f_n(x) vale 1 esattamente una volta per ciclo, 0 per tutti gli altri blocchi dello stesso ciclo")
print("-> essendo i cicli infiniti, f_n(x)=1 infinite volte e f_n(x)=0 infinite volte: NON converge")
```

**1.** Fissato $\varepsilon>0$: per la disuguaglianza triangolare, $|f(x)-g(x)|\le|f(x)-f_n(x)|+|f_n(x)-g(x)|$. Se $|f(x)-g(x)|\ge\varepsilon$, allora almeno uno dei due termini a destra deve essere $\ge\varepsilon/2$ (altrimenti la somma sarebbe $<\varepsilon$): $\{x:|f(x)-g(x)|\ge\varepsilon\}\subseteq\{x:|f_n(x)-f(x)|\ge\varepsilon/2\}\cup\{x:|f_n(x)-g(x)|\ge\varepsilon/2\}$. Per subadditività (notebook 13.2): $\mu(\{x:|f(x)-g(x)|\ge\varepsilon\})\le\mu(\{x:|f_n(x)-f(x)|\ge\varepsilon/2\})+\mu(\{x:|f_n(x)-g(x)|\ge\varepsilon/2\})$. Il membro destro tende a $0+0=0$ quando $n\to\infty$ (per le due ipotesi di convergenza in misura), ma il membro sinistro **non dipende da $n$**: dunque deve essere $\mu(\{x:|f(x)-g(x)|\ge\varepsilon\})=0$ per ogni $\varepsilon>0$. Prendendo $\varepsilon=1/k$ e usando la continuità dal basso (notebook 13.2, applicata a $\{x:|f(x)-g(x)|\ge1/k\}$ crescente al decrescere di $k$... più precisamente unendo su $k$): $\mu(\{x:f(x)\ne g(x)\})=\mu\big(\bigcup_k\{x:|f(x)-g(x)|\ge1/k\}\big)\le\sum_k\mu(\{x:|f(x)-g(x)|\ge1/k\})=\sum_k0=0$ (subadditività numerabile, esercizio del notebook 13.2). Dunque $f=g$ quasi ovunque.

**2.** $f_1$ ha larghezza $1$: $\mu(\{f_1\ge1/2\})=1$. $f_2,f_3$ hanno larghezza $1/2$: $\mu=0.5$ per ciascuno. $f_4,\ldots,f_7$ hanno larghezza $1/4$: $\mu=0.25$. $f_8$ (primo del ciclo successivo, larghezza $1/8$): $\mu=0.125$. In sintesi: $1,\ 0.5,\ 0.5,\ 0.25,\ 0.25,\ 0.25,\ 0.25,\ 0.125,\ldots$ — la successione delle misure tende chiaramente a $0$ (dimezzandosi a ogni ciclo completo).

**3.** Scegliendo, per ogni ciclo $k$, **un solo** blocco (per esempio sempre il primo, $[0,1/2^k)$) invece di tutti i $2^k$ blocchi del ciclo, si ottiene una sottosuccessione $f_{n_k}=\mathbb 1_{[0,1/2^k)}$. Per un punto fissato $x>0$: non appena $1/2^k<x$ (cioè $k$ abbastanza grande), il blocco $[0,1/2^k)$ non copre più $x$, quindi $f_{n_k}(x)=0$ definitivamente: $f_{n_k}(x)\to0$. Solo $x=0$ resta sempre coperto ($f_{n_k}(0)=1$ per ogni $k$), ma $\{0\}$ ha misura nulla: questa sottosuccessione converge a $0$ **quasi ovunque** (e persino ovunque tranne un solo punto).

**4.** Per convergenza uniforme (notebook 10.16), per ogni $\varepsilon>0$ esiste $N$ tale che $\sup_{x\in X}|f_n(x)-f(x)|<\varepsilon$ per $n\ge N$. Questo significa che $|f_n(x)-f(x)|<\varepsilon$ per **ogni** $x\in X$ (non solo per la maggior parte): dunque $\{x:|f_n(x)-f(x)|\ge\varepsilon\}=\emptyset$ per $n\ge N$. Essendo $\mu(\emptyset)=0$ (assioma di misura), $\mu(\{x:|f_n(x)-f(x)|\ge\varepsilon\})=0$ per $n\ge N$, quindi banalmente $\to0$. Dunque $f_n\to f$ in misura (l'ipotesi $\mu(X)<\infty$ non è nemmeno necessaria qui, a differenza del teorema principale della teoria che parte dalla convergenza q.o., più debole).

## Riepilogo finale

- La **convergenza in misura** richiede che l'insieme dove $|f_n-f|\ge\varepsilon$ abbia misura che tende a $0$, senza controllare cosa succede in ogni singolo punto.
- Su misura finita, la convergenza quasi ovunque **implica** quella in misura (via il teorema di Egorov), ma **non vale il viceversa**: la successione "a macchina da scrivere" converge in misura a $0$ pur non convergendo in **nessun** punto.
- Ogni successione che converge in misura ammette però una sottosuccessione che converge quasi ovunque: un fatto di consolazione che recupera parzialmente il legame tra le due nozioni.
- Le due convergenze (q.o. e in misura) catturano aspetti complementari: una guarda "cosa succede in ogni punto nel lungo periodo", l'altra guarda "quanto spazio occupa il disaccordo a ogni istante".
