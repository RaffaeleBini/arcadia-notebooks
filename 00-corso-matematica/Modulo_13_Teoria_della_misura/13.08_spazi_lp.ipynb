**Prerequisiti:** 13.7 Teoremi di convergenza, 6.20 Prodotti scalari, 12.8 Completezza (spazi metrici).


## Perché ci interessa

Nel notebook 6.20 abbiamo definito la norma euclidea $\|v\|=\sqrt{v\cdot v}$ su $\mathbb R^n$, misurando la "grandezza" di un vettore con $n$ componenti. Una funzione $f:X\to\mathbb R$ può essere pensata come un "vettore con infinite componenti" (un valore per ogni $x\in X$): ha senso allora chiedersi se esista una norma analoga per le funzioni. Gli **spazi $L^p$**, costruiti con l'integrale di Lebesgue (notebook 13.4), rispondono a questa domanda: forniscono una famiglia di norme su spazi di funzioni, ciascuna con proprietà diverse, che diventano lo scenario naturale per l'analisi funzionale (Modulo 14) e la teoria delle equazioni alle derivate parziali.

## Teoria, passo dopo passo

### Definizione di $L^p$

Per $p\ge1$, $L^p(X,\mu)$ è l'insieme delle funzioni misurabili $f:X\to\mathbb R$ (identificando funzioni uguali quasi ovunque, notebook 13.5 — un dettaglio tecnico essenziale, discusso sotto) con $\int|f|^p\,d\mu<\infty$, dotato della **norma** $\|f\|_p=\Big(\int|f|^p\,d\mu\Big)^{1/p}$.

**Perché identificare funzioni uguali q.o.** Se $f=g$ quasi ovunque, allora $\int|f-g|^p\,d\mu=0$ (per l'esercizio 3 del notebook 13.4, essendo $|f-g|^p=0$ quasi ovunque), quindi $\|f-g\|_p=0$. Ma una vera norma deve soddisfare $\|h\|=0\Rightarrow h=0$ (positività, notebook 6.20/12.7): se non identificassimo $f$ e $g$, $\|\cdot\|_p$ non sarebbe una norma ma solo una "seminorma". La soluzione standard è definire $L^p$ come l'insieme delle **classi di equivalenza** rispetto all'uguaglianza q.o. (esattamente come i laterali del notebook 11.3, o le classi di $\mathbb Z/n\mathbb Z$): con questa identificazione, $\|\cdot\|_p$ diventa una vera norma.

### La disuguaglianza di Hölder

**Teorema (Hölder).** Se $\frac1p+\frac1q=1$ (con $p,q>1$, detti **esponenti coniugati**), allora $\int|fg|\,d\mu\le\|f\|_p\|g\|_q$.

Per $p=q=2$, questa è **esattamente** la disuguaglianza di Cauchy-Schwarz del notebook 6.20, riscritta con l'integrale al posto della somma finita: $\int|fg|\,d\mu\le\Big(\int f^2\,d\mu\Big)^{1/2}\Big(\int g^2\,d\mu\Big)^{1/2}$ — lo stesso principio geometrico ("il prodotto scalare è limitato dal prodotto delle norme"), esteso da vettori finiti a funzioni. La dimostrazione generale usa la disuguaglianza di Young ($ab\le\frac{a^p}p+\frac{b^q}q$ per $a,b\ge0$, conseguenza della concavità del logaritmo, un fatto di analisi del Modulo 5) applicata a $a=|f(x)|/\|f\|_p$, $b=|g(x)|/\|g\|_q$, e poi integrando.

### La disuguaglianza di Minkowski (disuguaglianza triangolare per $\|\cdot\|_p$)

**Teorema (Minkowski).** Per $p\ge1$: $\|f+g\|_p\le\|f\|_p+\|g\|_p$.

**Dimostrazione (schema, usando Hölder).** $\int|f+g|^p\,d\mu=\int|f+g|\cdot|f+g|^{p-1}\,d\mu\le\int|f|\cdot|f+g|^{p-1}\,d\mu+\int|g|\cdot|f+g|^{p-1}\,d\mu$ (disuguaglianza triangolare puntuale $|f+g|\le|f|+|g|$, poi integrata). Applicando Hölder a ciascun termine (con $q=p/(p-1)$, l'esponente coniugato di $p$): $\int|f|\cdot|f+g|^{p-1}\,d\mu\le\|f\|_p\cdot\||f+g|^{p-1}\|_q=\|f\|_p\cdot\|f+g\|_p^{p-1}$ (usando $(p-1)q=p$, verificabile dall'identità $1/p+1/q=1$). Sommando i due termini analoghi e dividendo per $\|f+g\|_p^{p-1}$ (quando non nullo), si ottiene $\|f+g\|_p\le\|f\|_p+\|g\|_p$. $\blacksquare$ Questa disuguaglianza è precisamente ciò che rende $\|\cdot\|_p$ una vera **norma** su $L^p$ (insieme a positività e omogeneità, verifiche immediate dalla definizione): $L^p$ è uno **spazio normato**, e quindi (notebook 12.7) uno spazio metrico con $d(f,g)=\|f-g\|_p$.

### Il teorema di Riesz-Fischer: $L^p$ è completo

**Teorema (Riesz-Fischer, enunciato).** $(L^p(X,\mu),\|\cdot\|_p)$ è uno spazio metrico **completo** (notebook 12.8) per ogni $p\ge1$: ogni successione di Cauchy in $L^p$ converge (in norma $L^p$) a un elemento di $L^p$.

Uno spazio normato completo si chiama **spazio di Banach** (concetto formale del notebook 14.2): $L^p$ è l'esempio archetipico. La dimostrazione (fuori portata qui) usa in modo essenziale il teorema della convergenza dominata (notebook 13.7): data una successione di Cauchy, si estrae una sottosuccessione che converge quasi ovunque abbastanza rapidamente da essere dominata, permettendo di concludere la convergenza in norma. Questo è un esempio importante di come i teoremi di convergenza della teoria della misura siano lo strumento tecnico dietro risultati strutturali dell'analisi funzionale.

```python
import numpy as np
from scipy import integrate

# Verifichiamo numericamente la disuguaglianza di Holder per p=q=2 (Cauchy-Schwarz) su un esempio concreto
def f(x):
    return np.sin(x)

def g(x):
    return np.exp(-x)

dominio = (0, 5)

norma_f_2 = np.sqrt(integrate.quad(lambda x: f(x)**2, *dominio)[0])
norma_g_2 = np.sqrt(integrate.quad(lambda x: g(x)**2, *dominio)[0])
integrale_prodotto = integrate.quad(lambda x: abs(f(x) * g(x)), *dominio)[0]

print(f"||f||_2 = {norma_f_2:.5f}")
print(f"||g||_2 = {norma_g_2:.5f}")
print(f"integrale |f*g| = {integrale_prodotto:.5f}")
print(f"||f||_2 * ||g||_2 = {norma_f_2 * norma_g_2:.5f}")
print(f"Holder (Cauchy-Schwarz) verificata: {integrale_prodotto:.5f} <= {norma_f_2*norma_g_2:.5f}? {integrale_prodotto <= norma_f_2*norma_g_2}")

# Verifica di Minkowski
norma_somma_2 = np.sqrt(integrate.quad(lambda x: (f(x)+g(x))**2, *dominio)[0])
print(f"\n||f+g||_2 = {norma_somma_2:.5f}")
print(f"||f||_2 + ||g||_2 = {norma_f_2 + norma_g_2:.5f}")
print(f"Minkowski verificata: {norma_somma_2:.5f} <= {norma_f_2+norma_g_2:.5f}? {norma_somma_2 <= norma_f_2+norma_g_2}")
```

## Esercizio svolto e commentato

**Problema:** calcola $\|f\|_p$ per $f(x)=1$ su $[0,1]$, $\|f\|_p$ per $f(x)=x$ su $[0,1]$, e mostra che $\|f\|_p\le\|f\|_q$ per $p\le q$ quando $\mu(X)=1$ (come qui), verificandolo su questi esempi.

**Ragionamento:**

**Passo 1 — calcoliamo $\|1\|_p$.** $\|1\|_p=\Big(\int_0^1 1^p\,dx\Big)^{1/p}=(1)^{1/p}=1$, per **ogni** $p\ge1$ (la funzione costante $1$ ha norma $1$ in ogni $L^p$ su uno spazio di misura totale $1$).

**Passo 2 — calcoliamo $\|x\|_p$.** $\|x\|_p=\Big(\int_0^1x^p\,dx\Big)^{1/p}=\Big(\dfrac{1}{p+1}\Big)^{1/p}$.

**Passo 3 — verifichiamo la monotonia $\|f\|_p\le\|f\|_q$ per $p\le q$ sull'esempio $f=x$.** Confrontiamo $\|x\|_1=\dfrac12$ e $\|x\|_2=\Big(\dfrac13\Big)^{1/2}=\dfrac1{\sqrt3}\approx0.577$: infatti $0.5\le0.577$. Confrontiamo $\|x\|_2\approx0.577$ e $\|x\|_3=\Big(\dfrac14\Big)^{1/3}\approx0.63$: infatti $0.577\le0.63$. La disuguaglianza è verificata su questi valori.

**Passo 4 — spieghiamo perché vale in generale (quando $\mu(X)=1$).** Per $p\le q$, applicando Hölder a $|f|^p\cdot1$ con esponenti coniugati $q/p$ e il suo coniugato: $\int|f|^p\,d\mu\le\Big(\int|f|^{p\cdot(q/p)}\,d\mu\Big)^{p/q}\cdot\mu(X)^{1-p/q}=\Big(\int|f|^q\,d\mu\Big)^{p/q}$ (usando $\mu(X)=1$, quindi $\mu(X)^{1-p/q}=1$). Elevando entrambi i membri a $1/p$: $\|f\|_p\le\|f\|_q$.

**Conclusione:** su uno spazio di misura totale $1$ (come una misura di probabilità, o $[0,1]$ con Lebesgue), le norme $L^p$ sono **ordinate**: $\|f\|_1\le\|f\|_2\le\|f\|_3\le\cdots$, in particolare implicando che $L^q\subseteq L^p$ per $p\le q$ su spazi di misura finita — una proprietà che **fallisce** su spazi di misura infinita come $\mathbb R$ intero (dove nessuna delle due inclusioni vale in generale, un fatto lasciato come spunto di riflessione, non trattato negli esercizi).

```python
import numpy as np

def norma_p(potenza, p):
    # ||x^potenza||_p su [0,1] = (1/(potenza*p+1))^(1/p)
    return (1 / (potenza * p + 1)) ** (1/p)

print("Norme L^p di f(x)=x su [0,1]:")
for p in [1, 2, 3, 5, 10]:
    print(f"||x||_{p} = {norma_p(1, p):.5f}")

print("\nVerifica della monotonia crescente in p (su spazio di misura 1):")
valori = [norma_p(1, p) for p in [1, 2, 3, 5, 10]]
print("Sequenza crescente?", all(valori[i] <= valori[i+1] for i in range(len(valori)-1)))
```

## Riepilogo finale

- Gli **spazi $L^p$** raccolgono le funzioni (classi di equivalenza quasi ovunque) con $\int|f|^p\,d\mu<\infty$, dotate della norma $\|f\|_p=\big(\int|f|^p\,d\mu\big)^{1/p}$: la generalizzazione a spazi di funzioni della norma euclidea (notebook 6.20).
- Le disuguaglianze di **Hölder** (generalizzazione di Cauchy-Schwarz) e **Minkowski** (disuguaglianza triangolare) rendono $\|\cdot\|_p$ una vera norma.
- Il teorema di **Riesz-Fischer** garantisce che $L^p$ è **completo** (uno spazio di Banach, notebook 12.8/14.2): fondamentale per l'analisi funzionale, e dimostrato usando i teoremi di convergenza del notebook 13.7.
- Su spazi di misura finita, le norme $L^p$ sono ordinate ($\|f\|_p\le\|f\|_q$ per $p\le q$), ma su domini con singolarità o su $\mathbb R$ intero, nessuna inclusione generale vale tra $L^p$ diversi.
