# 13.5 — Convergenza quasi ovunque

**Prerequisiti:** 13.4 Integrale di Lebesgue, 10.16 Convergenza uniforme.


## Perché ci interessa

Nell'esercizio 3 del notebook 13.4 abbiamo dimostrato che una funzione nulla **quasi ovunque** ha integrale nullo: la teoria della misura è "cieca" rispetto a ciò che accade su un insieme di misura zero. Questo suggerisce che anche le nozioni di convergenza dovrebbero essere aggiornate: la convergenza puntuale ovunque (notebook 10.16) è spesso una richiesta più forte di quanto serva. La **convergenza quasi ovunque** indebolisce la convergenza puntuale esattamente della quantità giusta — ignorando insiemi di misura nulla — e si rivela la nozione naturale per i teoremi di passaggio al limite sotto il segno di integrale, che costruiremo nel notebook 13.7.

## Teoria, passo dopo passo

### Definizione di "quasi ovunque"

Una proprietà $P(x)$ vale **quasi ovunque** (abbreviato q.o., o a.e. dall'inglese "almost everywhere") rispetto a una misura $\mu$ se l'insieme $\{x:P(x)\text{ è falsa}\}$ ha misura $\mu$ nulla. Per esempio: "$f=g$ quasi ovunque" significa $\mu(\{x:f(x)\ne g(x)\})=0$.

### Convergenza quasi ovunque

Una successione di funzioni misurabili $(f_n)$ **converge quasi ovunque** a $f$ (scritto $f_n\to f$ q.o.) se $\mu(\{x:f_n(x)\not\to f(x)\})=0$: cioè, la convergenza puntuale (notebook 10.16, versione puntuale prima di quella uniforme) può fallire, ma solo su un insieme "trascurabile".

**Confronto con la convergenza uniforme (notebook 10.16).** La convergenza uniforme implica quella puntuale ovunque, che implica quella quasi ovunque: sono richieste via via più deboli. La convergenza quasi ovunque è **strettamente più debole**: esistono successioni che convergono q.o. ma non puntualmente ovunque, come vedremo nell'esempio seguente.

### Esempio: convergenza q.o. ma non ovunque

Su $[0,1]$, sia $f_n=\mathbb 1_{\{1/n\}}$ (la funzione indicatrice del singolo punto $1/n$). Per ogni $x\ne0$: se $x\ne1/n$ per ogni $n$ (il caso tipico), $f_n(x)=0$ per ogni $n$, quindi $f_n(x)\to0$; se $x=1/n_0$ per un particolare $n_0$, allora $f_{n_0}(x)=1$ ma $f_n(x)=0$ per $n\ne n_0$ — quindi comunque $f_n(x)\to0$ (essendo $f_n(x)=0$ per tutti gli $n$ tranne al più uno). In $x=0$: $f_n(0)=0$ per ogni $n$ (essendo $1/n\ne0$), quindi anche qui $f_n(0)\to0$. **In realtà questa successione converge puntualmente ovunque a $0$** — un esempio più istruttivo segue: sia $g_n=n\cdot\mathbb 1_{[0,1/n]}$. Per $x=0$: $g_n(0)=n\to\infty$, **non converge**. Per $x>0$: appena $n>1/x$, si ha $1/n<x$, quindi $g_n(x)=0$ per tutti gli $n$ abbastanza grandi, cioè $g_n(x)\to0$. Dunque $g_n\to0$ su $(0,1]$ (ovunque tranne $\{0\}$), e $\{0\}$ ha misura di Lebesgue nulla (notebook 13.2): $g_n\to0$ **quasi ovunque**, ma non ovunque (fallisce esattamente in $x=0$).

### La funzione "spazzaneve" (typewriter sequence): convergenza q.o. senza convergenza in nessun punto fisso... ma non è questo il caso

Un esempio più drammatico, usato nel notebook 13.6 per contrasto con la convergenza in misura: sulla successione "a macchina da scrivere" definita spostando ripetutamente un blocco di larghezza decrescente lungo $[0,1]$, si ha convergenza a $0$ in **quasi ogni** punto fissato (con un argomento di misura: la probabilità che un dato $x$ sia coperto da blocchi sempre più stretti tende a $0$), un fenomeno che analizzeremo con precisione nel prossimo notebook.

### Il teorema di Egorov (enunciato)

**Teorema (Egorov, enunciato senza dimostrazione completa).** Se $\mu(X)<\infty$ e $f_n\to f$ quasi ovunque, allora per ogni $\varepsilon>0$ esiste un insieme misurabile $E$ con $\mu(X\setminus E)<\varepsilon$ tale che $f_n\to f$ **uniformemente** su $E$.

Questo teorema è sorprendente: dice che la convergenza q.o. (debole) "diventa" convergenza uniforme (forte, notebook 10.16) se siamo disposti a scartare un insieme di misura arbitrariamente piccola. È un ponte tra la nozione debole e quella forte, valido solo quando la misura totale è finita (essenziale: il controesempio $g_n=n\cdot\mathbb 1_{[0,1/n]}$ sopra ha dominio $[0,1]$ con $\lambda([0,1])=1<\infty$, e infatti la convergenza uniforme si recupera scartando un piccolo intorno di $0$).

```python
import numpy as np
import matplotlib.pyplot as plt

def g_n(x, n):
    return np.where((x >= 0) & (x <= 1/n), n, 0)

x = np.linspace(0, 1, 1000)
fig, ax = plt.subplots(figsize=(8, 5))
for n in [2, 5, 20, 100]:
    ax.plot(x, g_n(x, n), label=f"g_{n}(x)")

ax.set_title("g_n(x) = n*1_[0,1/n](x): converge a 0 quasi ovunque, ma non in x=0")
ax.legend()
ax.set_ylim(0, 15)
plt.tight_layout()
plt.show()

# Verifica numerica: per x>0 fissato, g_n(x)->0; per x=0, g_n(0)=n->infinito
for x_test in [0.5, 0.1, 0.01]:
    valori = [g_n(np.array([x_test]), n)[0] for n in [1, 10, 100, 1000]]
    print(f"x={x_test}: g_n(x) per n=1,10,100,1000 -> {valori}")

print(f"\nx=0: g_n(0) per n=1,10,100,1000 -> {[g_n(np.array([0.0]), n)[0] for n in [1,10,100,1000]]} (diverge)")
print("Ma {0} ha misura di Lebesgue nulla: g_n -> 0 quasi ovunque")
```

## Esercizio svolto e commentato

**Problema:** dimostra che se $f_n\to f$ q.o. e $f_n\to h$ q.o. (rispetto alla stessa misura $\mu$), allora $f=h$ quasi ovunque (unicità del limite q.o.).

**Ragionamento:**

**Passo 1 — identifichiamo gli insiemi "cattivi".** Sia $N_1=\{x:f_n(x)\not\to f(x)\}$ (con $\mu(N_1)=0$, per ipotesi $f_n\to f$ q.o.) e $N_2=\{x:f_n(x)\not\to h(x)\}$ (con $\mu(N_2)=0$, per ipotesi $f_n\to h$ q.o.).

**Passo 2 — fuori da $N_1\cup N_2$, entrambi i limiti esistono e coincidono.** Se $x\notin N_1\cup N_2$, allora $x\notin N_1$ (quindi $f_n(x)\to f(x)$) e $x\notin N_2$ (quindi $f_n(x)\to h(x)$). Per **unicità del limite di una successione numerica** (dimostrata rigorosamente nel Modulo 10, per successioni reali), essendo $f_n(x)$ un'unica successione di numeri reali con due limiti $f(x)$ e $h(x)$, deve essere $f(x)=h(x)$.

**Passo 3 — stimiamo la misura dell'insieme dove $f\ne h$.** $\{x:f(x)\ne h(x)\}\subseteq N_1\cup N_2$ (per il passo 2, in contrapposizione: se $f(x)\ne h(x)$, allora $x$ deve stare in $N_1\cup N_2$, altrimenti per il passo 2 sarebbe $f(x)=h(x)$). Per monotonia e subadditività della misura (notebook 13.2): $\mu(\{x:f(x)\ne h(x)\})\le\mu(N_1\cup N_2)\le\mu(N_1)+\mu(N_2)=0+0=0$.

**Conclusione:** $\mu(\{x:f(x)\ne h(x)\})=0$, cioè $f=h$ quasi ovunque. **Punto metodologico:** questo esercizio mostra la tecnica standard per lavorare con proprietà "quasi ovunque" — identificare gli insiemi eccezionali (di misura nulla) di ciascuna ipotesi, unirli (restando di misura nulla per subadditività), e ragionare "puntualmente" fuori da questa unione, dove tutte le ipotesi valgono contemporaneamente.

```python
import numpy as np

# Verifica computazionale: due successioni che convergono q.o. a limiti diversi solo su un insieme "piccolo"
# f_n -> f=0 ovunque tranne un insieme finito di punti (misura nulla)
# h_n -> h=0 ovunque tranne un altro insieme finito di punti

punti_eccezionali_f = {0.3, 0.7}  # dove f "sbaglia" (simulato)
punti_eccezionali_h = {0.5}       # dove h "sbaglia"

def f(x):
    return 1.0 if round(x, 1) in punti_eccezionali_f else 0.0

def h(x):
    return 1.0 if round(x, 1) in punti_eccezionali_h else 0.0

campione = np.round(np.linspace(0, 1, 11), 1)
disaccordo = [x for x in campione if f(x) != h(x)]
print("Punti campione dove f e h differiscono:", disaccordo)
print(f"Insieme di disaccordo ha {len(disaccordo)} punti su {len(campione)} campionati")
print("Essendo un insieme finito (quindi numerabile), ha misura di Lebesgue nulla: f=h quasi ovunque")
```

**1.** Siano $N_1=\{x:f_n(x)\not\to f(x)\}$, $N_2=\{x:g_n(x)\not\to g(x)\}$, entrambi di misura nulla. Dunque $\{x:f_n(x)\not\to f(x)\}=\emptyset$, che ha misura $0$ (assioma 1 di misura, notebook 13.2): $f_n\to f$ quasi ovunque (banalmente, con insieme eccezionale vuoto).

**3.** $\sup_{x\in[0,1]}|g_n(x)-0|=\sup_x g_n(x)=n$ (il massimo di $g_n$, raggiunto su $[0,1/n]$, vale $n$). Essendo $n\to\infty$, $\sup_x|g_n(x)-0|\not\to0$: la convergenza non è uniforme (anzi, il sup diverge, non solo non tende a zero).

**4.** Prendiamo $E=[\delta,1]$ con $\delta=0.1$: $\lambda([0,1]\setminus E)=\lambda([0,\delta))=\delta=0.1$ — per avere $<0.1$ stretto, scegliamo $\delta=0.05$, dando $\lambda([0,1]\setminus E)=0.05<0.1$. Su $E=[0.05,1]$: per $n>1/0.05=20$, si ha $1/n<0.05$, quindi $[0,1/n]\cap E=\emptyset$ (l'intervallo dove $g_n\ne0$ non tocca più $E$), cioè $g_n(x)=0$ per ogni $x\in E$ quando $n>20$. Dunque $\sup_{x\in E}|g_n(x)-0|=0$ per $n>20$, che banalmente tende a $0$: $g_n\to0$ uniformemente su $E$, come garantito da Egorov.

## Riepilogo finale

- Una proprietà vale **quasi ovunque** se l'insieme dove fallisce ha misura nulla; la teoria della misura tratta come identiche due funzioni uguali q.o.
- La **convergenza quasi ovunque** indebolisce la convergenza puntuale ovunque (notebook 10.16), tollerando un fallimento su un insieme trascurabile — esempio: $g_n=n\cdot\mathbb 1_{[0,1/n]}\to0$ q.o. ma non in $x=0$.
- Il limite quasi ovunque è **unico a meno di insiemi di misura nulla**: se una successione converge q.o. a due funzioni diverse, queste devono coincidere q.o.
- Il **teorema di Egorov** collega convergenza q.o. e convergenza uniforme: su misura finita, la convergenza q.o. diventa uniforme scartando un insieme di misura arbitrariamente piccola.
