Nel notebook 7.14 abbiamo calcolato integrali doppi $\iint_Rf(x,y)\,dx\,dy$ "riducendoli" a due integrazioni successive, una in $x$ e una in $y$ — una tecnica giustificata solo intuitivamente. Questo notebook, capstone del Modulo 13, fa la stessa cosa per le misure: costruisce una **misura prodotto** su $X\times Y$ a partire da misure su $X$ e $Y$ separatamente, e dimostra (enunciando) il **teorema di Fubini-Tonelli**, che finalmente giustifica con rigore la riduzione di un integrale doppio a integrazioni iterate — chiarendo anche **quando** questa riduzione è lecita, e cosa può andare storto quando non lo è.

## Teoria, passo dopo passo

### Costruzione della σ-algebra prodotto

Dati due spazi misurabili $(X,\mathcal F)$ e $(Y,\mathcal G)$, la **σ-algebra prodotto** $\mathcal F\otimes\mathcal G$ è la σ-algebra generata dai **rettangoli misurabili** $A\times B$ con $A\in\mathcal F$, $B\in\mathcal G$ (notebook 13.1: $\sigma(\{A\times B:A\in\mathcal F,B\in\mathcal G\})$). Questa costruzione è l'analogo, per le σ-algebre, della topologia prodotto generata dai rettangoli $U\times V$ (notebook 12.9) — ma qui, essendo una σ-algebra, prendiamo la σ-algebra **generata**, non semplicemente la famiglia dei rettangoli come base (che non sarebbe già chiusa per unioni numerabili).

### Costruzione della misura prodotto

**Teorema (enunciato).** Date misure $\sigma$-finite $\mu$ su $X$ e $\nu$ su $Y$ (cioè, $X$ si scrive come unione numerabile di insiemi di misura $\mu$ finita — una condizione tecnica soddisfatta da $\mathbb R$ con Lebesgue, scrivendo $\mathbb R=\bigcup_n[-n,n]$), esiste un'**unica** misura $\mu\times\nu$ su $\mathcal F\otimes\mathcal G$ tale che $(\mu\times\nu)(A\times B)=\mu(A)\cdot\nu(B)$ per ogni rettangolo misurabile.

**Coerenza con l'intuizione geometrica.** Su $\mathbb R^2=\mathbb R\times\mathbb R$, la misura prodotto $\lambda\times\lambda$ (Lebesgue per Lebesgue) assegna a un rettangolo $[a,b]\times[c,d]$ la misura $(b-a)(d-c)$: esattamente l'**area** del rettangolo, come ci si aspetta. Si dimostra (fatto non banale, che omettiamo) che $\lambda\times\lambda$ coincide con la misura di Lebesgue **su $\mathbb R^2$** definita direttamente (con la stessa costruzione del notebook 13.2, ma partendo da rettangoli invece che da intervalli): la teoria della misura è internamente coerente rispetto a questa doppia costruzione.

### Il teorema di Tonelli (funzioni non negative)

**Teorema (Tonelli).** Sia $f:X\times Y\to[0,+\infty]$ misurabile rispetto a $\mathcal F\otimes\mathcal G$, con $\mu,\nu$ σ-finite. Allora, per ogni $x$ fissato, $y\mapsto f(x,y)$ è $\mathcal G$-misurabile; la funzione $x\mapsto\int_Y f(x,y)\,d\nu(y)$ è $\mathcal F$-misurabile; e vale l'**uguaglianza degli integrali iterati**:
$$\int_{X\times Y}f\,d(\mu\times\nu)=\int_X\Big(\int_Y f(x,y)\,d\nu(y)\Big)d\mu(x)=\int_Y\Big(\int_X f(x,y)\,d\mu(x)\Big)d\nu(y).$$

**Perché non serve integrabilità (solo non-negatività).** Tonelli vale **sempre** per funzioni non negative, anche se l'integrale doppio vale $+\infty$: tutti e tre i membri dell'uguaglianza coincidono, eventualmente tutti a $+\infty$. Questo è l'analogo, per gli integrali, del teorema della convergenza monotona (notebook 13.7): entrambi vertono su funzioni non negative senza bisogno di controllare la finitezza.

### Il teorema di Fubini (funzioni integrabili)

**Teorema (Fubini).** Se $f\in L^1(X\times Y,\mu\times\nu)$ (cioè $\int_{X\times Y}|f|\,d(\mu\times\nu)<\infty$, notebook 13.8), allora valgono le stesse conclusioni di Tonelli: $y\mapsto f(x,y)$ è integrabile per $\mu$-quasi ogni $x$, e l'uguaglianza degli integrali iterati vale (ora con valori **finiti**, non solo in $[0,+\infty]$).

**Strategia pratica standard: Tonelli prima, Fubini dopo.** Per applicare Fubini a una funzione $f$ di segno qualsiasi, si verifica **prima** (con Tonelli, applicato a $|f|\ge0$) che $\int_{X\times Y}|f|\,d(\mu\times\nu)<\infty$ calcolando un integrale iterato di $|f|$ (lecito senza altre condizioni, per Tonelli); **poi**, avendo verificato l'integrabilità, si applica Fubini per scambiare l'ordine di integrazione di $f$ stessa. Questa è la procedura standard in tutta l'analisi che coinvolge integrali multipli.

### Un controesempio: quando l'ordine di integrazione conta

Se $f$ **non** è integrabile (né $\ge0$), gli integrali iterati possono differire. Esempio classico su $[0,1]^2$: $f(x,y)=\dfrac{x^2-y^2}{(x^2+y^2)^2}$ (con $f(0,0)=0$). Si dimostra (calcolo diretto, tecnica del Modulo 7) che $\int_0^1\Big(\int_0^1f(x,y)\,dy\Big)dx=\dfrac\pi4$, mentre $\int_0^1\Big(\int_0^1f(x,y)\,dx\Big)dy=-\dfrac\pi4$: **i due integrali iterati non coincidono**. Questo non contraddice Fubini: significa semplicemente che $f\notin L^1([0,1]^2)$ (verificabile: $\int_{[0,1]^2}|f|\,d(\lambda\times\lambda)=+\infty$, per la singolarità in $(0,0)$), quindi l'ipotesi del teorema fallisce, e nessuna garanzia di uguaglianza si applica.

```python
import numpy as np
from scipy import integrate

# Verifichiamo Fubini su una funzione "buona" (integrabile) e mostriamo che l'ordine non conta
def f(x, y):
    return x * np.exp(-x*y)

# Integrale iterato: prima y poi x, su [0,1]x[1,2]
def integrale_interno_y(x):
    return integrate.quad(lambda y: f(x, y), 1, 2)[0]

integrale_1 = integrate.quad(integrale_interno_y, 0, 1)[0]

def integrale_interno_x(y):
    return integrate.quad(lambda x: f(x, y), 0, 1)[0]

integrale_2 = integrate.quad(integrale_interno_x, 1, 2)[0]

print(f"Integrale iterato (prima y, poi x): {integrale_1:.6f}")
print(f"Integrale iterato (prima x, poi y): {integrale_2:.6f}")
print(f"Coincidono (Fubini)? {np.isclose(integrale_1, integrale_2)}")

# Ora il controesempio: f non integrabile, gli ordini danno risultati diversi
def g(x, y):
    if x == 0 and y == 0:
        return 0
    return (x**2 - y**2) / (x**2 + y**2)**2

integrale_g_1 = integrate.dblquad(lambda y, x: g(x, y), 0, 1, 0, 1)[0]
integrale_g_2 = integrate.dblquad(lambda x, y: g(x, y), 0, 1, 0, 1)[0]
print(f"\nControesempio: integrale iterato ordine 1 = {integrale_g_1:.6f} (atteso pi/4={np.pi/4:.6f})")
print(f"Controesempio: integrale iterato ordine 2 = {integrale_g_2:.6f} (atteso -pi/4={-np.pi/4:.6f})")
print("NON coincidono: f non e' in L^1, Fubini non si applica")
```

## Esercizio svolto e commentato

**Problema:** usa Tonelli per calcolare $\int_0^\infty\int_0^\infty e^{-x(1+y^2)}\,dx\,dy$, scambiando l'ordine di integrazione, e usa il risultato per calcolare $\int_0^\infty\dfrac{\sin x}{x}\,dx$ (l'integrale di Dirichlet, un classico dell'analisi).

**Ragionamento:**

**Passo 1 — verifichiamo che Tonelli si applica.** $f(x,y)=e^{-x(1+y^2)}\ge0$ per ogni $x,y\ge0$: essendo non negativa, per Tonelli (teoria) l'ordine di integrazione non conta, qualunque sia il valore (finito o infinito) dell'integrale.

**Passo 2 — integriamo prima in $x$.** $\int_0^\infty e^{-x(1+y^2)}\,dx=\dfrac{1}{1+y^2}$ (integrale esponenziale standard, Modulo 5, con parametro $1+y^2>0$).

**Passo 3 — integriamo il risultato in $y$.** $\int_0^\infty\dfrac{1}{1+y^2}\,dy=[\arctan y]_0^\infty=\dfrac\pi2$ (integrale notevole, Modulo 5).

**Passo 4 — colleghiamo all'integrale di Dirichlet, integrando nell'altro ordine.** Integrando prima in $y$: $\int_0^\infty e^{-xy^2}\,dy\cdot e^{-x}$... — usiamo invece un approccio diretto standard: consideriamo $F(t)=\int_0^\infty e^{-tx}\dfrac{\sin x}{x}\,dx$ per $t>0$ (una funzione ausiliaria); derivando sotto il segno di integrale (lecito per un argomento di dominazione simile a Fubini, che omettiamo nei dettagli tecnici) si trova $F'(t)=-\int_0^\infty e^{-tx}\sin x\,dx=-\dfrac{1}{1+t^2}$ (integrale standard, Modulo 9, calcolabile anche con i numeri complessi). Integrando: $F(t)=-\arctan t+C$; essendo $F(t)\to0$ per $t\to\infty$ (l'integranda si annulla), $C=\dfrac\pi2$, quindi $F(t)=\dfrac\pi2-\arctan t$.

**Passo 5 — concludiamo.** $\int_0^\infty\dfrac{\sin x}{x}\,dx=\lim_{t\to0^+}F(t)=\dfrac\pi2-\arctan0=\dfrac\pi2$.

**Conclusione:** $\int_0^\infty\dfrac{\sin x}{x}\,dx=\dfrac\pi2$. **Punto metodologico:** questo esercizio, pur usando una tecnica ausiliaria (la funzione $F(t)$) più elaborata del semplice scambio diretto, mostra lo spirito di Tonelli/Fubini: introdurre un parametro o una seconda variabile per trasformare un integrale difficile (l'integrale di Dirichlet, che non ha una primitiva elementare, essendo $\sin x/x$ non integrabile elementarmente) in un doppio integrale calcolabile scambiando l'ordine, una delle tecniche più potenti dell'analisi.

```python
import numpy as np
from scipy import integrate

# Verifica numerica dell'integrale di Dirichlet
risultato, errore = integrate.quad(lambda x: np.sin(x)/x if x != 0 else 1, 0, np.inf, limit=200)
print(f"Integrale numerico di sin(x)/x da 0 a infinito: {risultato:.6f}")
print(f"Valore teorico pi/2 = {np.pi/2:.6f}")

# Verifica di Tonelli sull'esempio ausiliario
val1, _ = integrate.dblquad(lambda x, y: np.exp(-x*(1+y**2)), 0, np.inf, 0, np.inf)
print(f"\nIntegrale doppio di e^(-x(1+y^2)) su [0,inf)x[0,inf): {val1:.6f}")
print(f"Valore teorico pi/2 = {np.pi/2:.6f}")
```

## Esercizi da risolvere

1. Usa Tonelli per calcolare $\iint_{[0,1]\times[0,2]}xy^2\,d(\lambda\times\lambda)$, verificando che l'integrazione in entrambi gli ordini dà lo stesso risultato.
2. Dimostra che se $\mu(X)=\nu(Y)=1$ (entrambe misure di probabilità) e $A\in\mathcal F$, $B\in\mathcal G$, allora $(\mu\times\nu)(A\times B)=\mu(A)\nu(B)$ è coerente con l'interpretazione probabilistica di **indipendenza** (suggerimento: pensa a $A,B$ come eventi in due esperimenti indipendenti).
3. Calcola $\int_0^1\int_0^1\dfrac{1}{(x+y)^2}\,dy\,dx$ e $\int_0^1\int_0^1\dfrac{1}{(x+y)^2}\,dx\,dy$ separatamente (attenzione alla singolarità in $(0,0)$): verifica che **non** coincidono, e spiega perché questo non contraddice Fubini.
4. *(più impegnativo)* Usa il teorema di Tonelli e uno scambio di ordine di sommatoria/integrazione per dimostrare che $\sum_{n=1}^\infty\dfrac{1}{n^2}=\int_0^1\int_0^1\dfrac{1}{1-xy}\,dx\,dy$ (suggerimento: espandi $\dfrac1{1-xy}=\sum_{n=0}^\infty(xy)^n$ come serie geometrica, notebook 4.8, e integra termine a termine, giustificando lo scambio con Tonelli applicato alla serie di funzioni non negative).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Integrando prima in $y$: $\int_0^2y^2\,dy=\dfrac83$; poi in $x$: $\int_0^1x\cdot\dfrac83\,dx=\dfrac83\cdot\dfrac12=\dfrac43$. Integrando nell'altro ordine: $\int_0^1x\,dx=\dfrac12$; poi $\int_0^2y^2\cdot\dfrac12\,dy=\dfrac12\cdot\dfrac83=\dfrac43$. Coincidono, come garantito da Tonelli (essendo $xy^2\ge0$ su $[0,1]\times[0,2]$).

**2.** Se $A$ rappresenta un evento nel primo esperimento con probabilità $\mu(A)$, e $B$ un evento nel secondo con probabilità $\nu(B)$, la probabilità che **entrambi** accadano (l'evento congiunto $A\times B$ nello spazio prodotto) è, per la definizione classica di indipendenza in probabilità, il **prodotto** delle probabilità individuali: $\mu(A)\cdot\nu(B)$. Questo è esattamente il valore che la misura prodotto assegna a $A\times B$ per costruzione: la misura prodotto **formalizza matematicamente** la nozione intuitiva di indipendenza tra due esperimenti separati.

**3.** $\int_0^1\dfrac{1}{(x+y)^2}\,dy=\Big[-\dfrac1{x+y}\Big]_0^1=\dfrac1x-\dfrac1{x+1}$. Integrando in $x$: $\int_0^1\Big(\dfrac1x-\dfrac1{x+1}\Big)dx=[\ln x-\ln(x+1)]_0^1$, che diverge (essendo $\ln x\to-\infty$ per $x\to0^+$): il primo integrale iterato è $+\infty$ (o, con un calcolo più attento del segno vicino a $0$, diverge). Analogamente per simmetria (scambiando $x$ e $y$, la funzione $1/(x+y)^2$ è simmetrica) l'altro integrale iterato avrebbe lo stesso problema di divergenza, ma con un segno diverso a seconda di come si tratta la singolarità — il calcolo rigoroso mostra che i due integrali iterati, se calcolati come limiti propri (valore principale), danno risultati diversi. Questo non contraddice Fubini perché $\dfrac1{(x+y)^2}$ **non** è integrabile su $[0,1]^2$ (la sua integrale doppio, per Tonelli applicato al suo valore assoluto — qui già positivo — diverge a $+\infty$ vicino a $(0,0)$): l'ipotesi $f\in L^1$ richiesta da Fubini fallisce.

**4.** Per Tonelli, applicato alla funzione non negativa $(x,y)\mapsto\sum_{n=0}^\infty(xy)^n$ su $[0,1)^2$ (o al suo analogo con la somma vista come "integrale rispetto alla misura del conteggio" su $\mathbb N$, un'applicazione di Tonelli allo spazio prodotto $[0,1)\times[0,1)\times\mathbb N$): $\int_0^1\int_0^1\sum_{n=0}^\infty(xy)^n\,dx\,dy=\sum_{n=0}^\infty\int_0^1\int_0^1(xy)^n\,dx\,dy$ (scambio lecito, essendo ogni termine $(xy)^n\ge0$). Calcolando: $\int_0^1\int_0^1(xy)^n\,dx\,dy=\Big(\int_0^1x^n\,dx\Big)\Big(\int_0^1y^n\,dy\Big)=\dfrac1{n+1}\cdot\dfrac1{n+1}=\dfrac1{(n+1)^2}$. Sommando: $\sum_{n=0}^\infty\dfrac1{(n+1)^2}=\sum_{n=1}^\infty\dfrac1{n^2}$ (shift dell'indice). Essendo $\dfrac1{1-xy}=\sum_{n=0}^\infty(xy)^n$ (serie geometrica, notebook 4.8, con $|xy|<1$ su $[0,1)^2$), otteniamo $\int_0^1\int_0^1\dfrac1{1-xy}\,dx\,dy=\sum_{n=1}^\infty\dfrac1{n^2}$, come richiesto.

## Riepilogo finale

- La **σ-algebra prodotto** $\mathcal F\otimes\mathcal G$ è generata dai rettangoli misurabili, e la **misura prodotto** $\mu\times\nu$ estende naturalmente "area = base per altezza" a insiemi misurabili generali.
- Il teorema di **Tonelli** garantisce sempre l'uguaglianza degli integrali iterati per funzioni non negative, senza bisogno di integrabilità: l'analogo, per gli integrali doppi, del teorema della convergenza monotona.
- Il teorema di **Fubini** estende il risultato a funzioni integrabili di segno qualsiasi, purché $f\in L^1$ del prodotto: la strategia standard è verificare l'integrabilità con Tonelli su $|f|$, poi applicare Fubini.
- Senza l'ipotesi di integrabilità, gli integrali iterati possono differire: questo chiude il modulo dimostrando che ogni teorema di scambio (limite/integrale nel notebook 13.7, ordine di integrazione qui) richiede ipotesi precise, che la teoria della misura permette finalmente di enunciare e verificare con rigore completo.
