## Perché ci interessa

Nel notebook 13.1 abbiamo stabilito **su quali insiemi** ha senso parlare di lunghezza/area/probabilità — gli insiemi misurabili di una σ-algebra. Ora serve la funzione stessa che assegna il "peso" numerico: la **misura**. Vogliamo che generalizzi la lunghezza di un intervallo ($\ell([a,b])=b-a$) mantenendo la proprietà più naturale che ci si aspetta: se spezziamo un insieme in pezzi numerabili disgiunti, la misura del tutto è la somma delle misure dei pezzi — una versione infinita, ma numerabile, dell'ovvia proprietà "l'area di due rettangoli disgiunti è la somma delle loro aree".

## Teoria, passo dopo passo

### Definizione di misura

Sia $(X,\mathcal F)$ uno spazio misurabile (notebook 13.1). Una **misura** è una funzione $\mu:\mathcal F\to[0,+\infty]$ (valori non negativi, eventualmente $+\infty$) tale che:

1. $\mu(\emptyset)=0$;
2. **σ-additività (numerabile)**: se $A_1,A_2,\ldots\in\mathcal F$ sono a due a due **disgiunti**, allora $\mu\Big(\bigcup_{n=1}^\infty A_n\Big)=\sum_{n=1}^\infty\mu(A_n)$ (la serie a destra, di termini non negativi, converge sempre a un valore in $[0,+\infty]$, notebook 4.7 — eventualmente diverge a $+\infty$, il che è ammesso).

La tripla $(X,\mathcal F,\mu)$ si chiama **spazio di misura**. Se $\mu(X)=1$, $\mu$ si dice **misura di probabilità** (collegamento diretto con la teoria della probabilità: gli insiemi misurabili sono gli "eventi", e $\mu$ è la probabilità).

### Proprietà elementari

**Monotonia.** Se $A\subseteq B$ (entrambi misurabili), allora $\mu(A)\le\mu(B)$.

**Dimostrazione.** $B=A\cup(B\setminus A)$, unione **disgiunta** (chiusura per differenza dimostrata nel notebook 13.1). Per σ-additività (con solo due insiemi, un caso particolare della somma numerabile, completando con insiemi vuoti): $\mu(B)=\mu(A)+\mu(B\setminus A)\ge\mu(A)$ (essendo $\mu(B\setminus A)\ge0$). $\blacksquare$

**Continuità dal basso.** Se $A_1\subseteq A_2\subseteq\cdots$ (successione crescente di misurabili), allora $\mu\Big(\bigcup_n A_n\Big)=\lim_{n\to\infty}\mu(A_n)$.

**Dimostrazione.** Poniamo $B_1=A_1$, $B_n=A_n\setminus A_{n-1}$ per $n\ge2$: gli $B_n$ sono a due a due disgiunti (per costruzione) e $\bigcup_{k=1}^n B_k=A_n$ (verifica diretta per induzione, notebook 10.4), quindi $\bigcup_n B_n=\bigcup_n A_n$. Per σ-additività: $\mu\Big(\bigcup_n A_n\Big)=\mu\Big(\bigcup_n B_n\Big)=\sum_{n=1}^\infty\mu(B_n)=\lim_{N\to\infty}\sum_{n=1}^N\mu(B_n)$ (definizione di somma di una serie come limite delle somme parziali, notebook 4.7) $=\lim_{N\to\infty}\mu(A_N)$ (essendo $\sum_{n=1}^N\mu(B_n)=\mu(A_N)$ per lo stesso argomento usato per l'unione, applicato all'unione finita $A_N=\bigcup_{k=1}^N B_k$). $\blacksquare$

Questa proprietà è l'analogo, per le misure, del teorema sulle successioni monotone limitate (notebook 4.4, 10.10): la misura di un'unione crescente è sempre il limite delle misure, senza bisogno di verificare altre condizioni.

### La misura di Lebesgue su $\mathbb R$ (costruzione, cenni)

La **misura di Lebesgue** $\lambda$ estende la lunghezza degli intervalli, $\lambda([a,b])=b-a$, a tutta la σ-algebra di Borel $\mathcal B(\mathbb R)$ (e oltre, alla sua completazione — cenno che omettiamo). La costruzione rigorosa passa per la **misura esterna** $\lambda^*(A)=\inf\Big\{\sum_n(b_n-a_n):A\subseteq\bigcup_n(a_n,b_n)\Big\}$ (l'estremo inferiore, che esiste per completezza, notebook 10.9, essendo l'insieme dei possibili valori limitato inferiormente da $0$), definita per **ogni** sottoinsieme di $\mathbb R$, seguita dalla verifica (teorema di Carathéodory, che esula da questo notebook introduttivo) che $\lambda^*$ ristretta ai boreliani soddisfa la σ-additività, diventando così una vera misura.

**Proprietà chiave, verificabili direttamente dalla definizione di misura esterna:** $\lambda(\{x\})=0$ per ogni punto (un singoletto si ricopre con intervalli $(x-\varepsilon/2,x+\varepsilon/2)$ di lunghezza $\varepsilon$ arbitrariamente piccola); per σ-additività, $\lambda(\mathbb Q)=0$ (unione numerabile di singoletti, notebook 10.5, ciascuno di misura $0$, quindi la somma della serie è $0$): un insieme **denso** in $\mathbb R$ può avere misura nulla, un fatto che sorprende chi pensa alla misura come "quanto spazio occupa visivamente".

```python
import numpy as np

# Approssimiamo numericamente la misura esterna di Lebesgue di [0,1] ricoprendolo con intervalli
# e verifichiamo che l'estremo inferiore delle lunghezze totali tende a 1 (la lunghezza "ovvia")

def ricopri_e_somma(a, b, n_intervalli):
    # ricopriamo [a,b] con n intervalli di uguale lunghezza, leggermente sovrapposti
    lunghezza_base = (b - a) / n_intervalli
    sovrapposizione = 0.001  # piccola sovrapposizione per garantire il ricoprimento
    lunghezza_totale = n_intervalli * (lunghezza_base + sovrapposizione)
    return lunghezza_totale

for n in [1, 5, 20, 100]:
    L = ricopri_e_somma(0, 1, n)
    print(f"Ricoprimento con {n} intervalli: lunghezza totale = {L:.6f} (tende a 1 = lambda([0,1]))")

# Verifica: la misura di un insieme numerabile (i primi 1000 razionali di [0,1]) e' arbitrariamente piccola
from fractions import Fraction
razionali = [Fraction(p, q) for q in range(1, 20) for p in range(0, q+1)]
razionali = sorted(set(r for r in razionali if 0 <= r <= 1))[:50]
print(f"\nNumero di razionali campionati: {len(razionali)}")
epsilon = 0.001
lunghezza_copertura = sum(epsilon / (2**i) for i in range(1, len(razionali)+1))
print(f"Ricoprendo l'n-esimo razionale con un intervallo di lunghezza epsilon/2^n:")
print(f"Lunghezza totale del ricoprimento: {lunghezza_copertura:.6f} (puo' essere resa piccola a piacere)")
```

## Esercizio svolto e commentato

**Problema:** dimostra che, per una misura $\mu$ su $(X,\mathcal F)$, vale la **subadditività numerabile**: per ogni successione $A_1,A_2,\ldots\in\mathcal F$ (non necessariamente disgiunti), $\mu\Big(\bigcup_n A_n\Big)\le\sum_n\mu(A_n)$.

**Ragionamento:**

**Passo 1 — il problema con insiemi non disgiunti.** La σ-additività (assioma 2) si applica solo a insiemi **disgiunti**: se gli $A_n$ si sovrappongono, sommare le loro misure "conta più volte" le parti comuni, quindi ci aspettiamo una **disuguaglianza**, non un'uguaglianza.

**Passo 2 — costruiamo una successione disgiunta con la stessa unione.** Definiamo $B_1=A_1$, $B_n=A_n\setminus(A_1\cup\cdots\cup A_{n-1})$ per $n\ge2$ (togliamo da ogni $A_n$ tutto ciò che è già stato "coperto" dai precedenti). Gli $B_n$ sono a due a due disgiunti (per costruzione: $B_n$ esclude esplicitamente $A_1,\ldots,A_{n-1}$, e $B_m\subseteq A_m$ per $m<n$), e $\bigcup_n B_n=\bigcup_n A_n$ (ogni punto della seconda unione appartiene a un primo $A_n$ che lo contiene, quindi appartiene al corrispondente $B_n$ — non essendo stato "tolto" da un $A_k$ precedente proprio perché $n$ è il primo indice, per definizione di $B_n$).

**Passo 3 — applichiamo σ-additività ai $B_n$ e confrontiamo con gli $A_n$.** Per σ-additività (essendo i $B_n$ disgiunti): $\mu\Big(\bigcup_n A_n\Big)=\mu\Big(\bigcup_n B_n\Big)=\sum_n\mu(B_n)$. Essendo $B_n\subseteq A_n$ per costruzione, per monotonia (teoria) $\mu(B_n)\le\mu(A_n)$ per ogni $n$. Sommando termine a termine (una serie di termini non negativi, dove ogni termine è minore o uguale al corrispondente termine di un'altra serie, ha somma minore o uguale, notebook 4.7): $\sum_n\mu(B_n)\le\sum_n\mu(A_n)$.

**Conclusione:** $\mu\Big(\bigcup_n A_n\Big)=\sum_n\mu(B_n)\le\sum_n\mu(A_n)$. **Applicazione pratica immediata:** questa disuguaglianza è ciò che garantisce $\lambda(\mathbb Q)=0$ nella teoria (senza dover verificare che i singoletti coprenti fossero disgiunti nella costruzione della misura esterna — la subadditività lo rende automatico anche per ricoprimenti che si sovrappongono).

```python
import numpy as np

def misura_intervallo(a, b):
    return max(0, b - a)

# Esempio concreto: due intervalli sovrapposti, confrontiamo mu(unione) con somma delle misure
A1 = (0, 3)
A2 = (2, 5)  # si sovrappone con A1 in (2,3)

# Unione = (0,5) (essendo gli intervalli sovrapposti e "connessi")
mu_unione = misura_intervallo(0, 5)
somma_misure = misura_intervallo(*A1) + misura_intervallo(*A2)

print(f"mu(A1) = {misura_intervallo(*A1)}, mu(A2) = {misura_intervallo(*A2)}")
print(f"mu(A1 U A2) = {mu_unione}")
print(f"mu(A1) + mu(A2) = {somma_misure}")
print(f"Subadditivita' verificata: mu(unione) <= somma? {mu_unione <= somma_misure}")
print(f"(La differenza {somma_misure - mu_unione} e' dovuta alla sovrapposizione (2,3), contata due volte a destra)")
```

Sottraendo $\mu(A_1)$ (finito) da entrambi i lati e cambiando segno: $\mu\Big(\bigcap_n A_n\Big)=\lim_n\mu(A_n)$.

**3.** Con $A_n=[n,+\infty)$ e $\mu=\lambda$ (Lebesgue): $A_n$ è decrescente, $\bigcap_n A_n=\emptyset$ (nessun numero reale sta in **tutti** i $[n,+\infty)$, altrimenti sarebbe $\ge n$ per ogni $n$, impossibile per l'Archimedeità di $\mathbb R$, notebook 10.9), quindi $\mu\Big(\bigcap_n A_n\Big)=\mu(\emptyset)=0$. Ma $\mu(A_n)=\lambda([n,+\infty))=+\infty$ per ogni $n$ (una semiretta ha lunghezza infinita), quindi $\lim_n\mu(A_n)=+\infty\ne0$. La continuità dall'alto fallisce, esattamente perché $\mu(A_1)=+\infty$ viola l'ipotesi di finitezza.

**4.** *Assioma 1*: $\mu(\emptyset)=|\emptyset|=0$. *Assioma 2*: siano $A_1,A_2,\ldots$ a due a due disgiunti. In ogni caso, la σ-additività è soddisfatta: la misura del conteggio è una misura.

## Riepilogo finale

- Una **misura** $\mu$ assegna a ogni insieme misurabile un valore in $[0,+\infty]$, con $\mu(\emptyset)=0$ e **σ-additività**: la misura di un'unione numerabile disgiunta è la somma delle misure.
- Da questi due assiomi seguono monotonia, subadditività numerabile, continuità dal basso (sempre) e dall'alto (se la prima misura è finita).
- La **misura di Lebesgue** estende la lunghezza degli intervalli a tutti i boreliani; ogni insieme numerabile (come $\mathbb Q$) ha misura di Lebesgue nulla, anche se denso in $\mathbb R$.
- La misura del conteggio e le misure di probabilità sono casi particolari della stessa struttura assiomatica: la teoria della misura unifica lunghezza, area, volume e probabilità sotto lo stesso linguaggio.
