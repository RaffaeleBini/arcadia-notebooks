## Perché ci interessa

Nel notebook 10.15 abbiamo costruito l'integrale di Riemann, basato sul concetto elementare di "lunghezza di un intervallo". Ma questa nozione di lunghezza si applica solo a insiemi molto semplici (intervalli, unioni finite di intervalli): non sappiamo dare un senso rigoroso alla "lunghezza" di $\mathbb Q\cap[0,1]$ (un insieme denso ma numerabile, notebook 10.5), né possiamo integrare funzioni come quella di Dirichlet (che vale $1$ sui razionali e $0$ sugli irrazionali), che non è Riemann-integrabile pur essendo limitata. La **teoria della misura** generalizza "lunghezza/area/volume" a una classe molto più ampia di insiemi. Il primo passo, indipendente dalla scelta della misura stessa, è stabilire **su quali insiemi** ha senso parlare di misura: questa è la **σ-algebra**, che raffina l'idea di topologia (notebook 12.1) richiedendo chiusura rispetto a operazioni **numerabili**, non solo finite.

## Teoria, passo dopo passo

### Perché la topologia non basta

Una topologia (notebook 12.1) è chiusa per unioni qualsiasi e intersezioni **finite**. Per misurare insiemi vogliamo poter fare operazioni **numerabili** (contare infiniti passi, ma non più che numerabili, notebook 10.5) sia di unione sia di intersezione sia di complementazione — perché tipiche costruzioni di analisi (come $\bigcap_n(-1/n,1/n)=\{0\}$, incontrata nel notebook 12.1) intrecciano tutte e tre le operazioni, e vogliamo che il risultato resti "misurabile". Serve una struttura diversa, su misura per queste esigenze.

### Definizione di σ-algebra

Una **σ-algebra** su un insieme $X$ è una famiglia $\mathcal F\subseteq\mathcal P(X)$ tale che:

1. $X\in\mathcal F$;
2. $\mathcal F$ è chiusa per complementazione: $A\in\mathcal F\Rightarrow X\setminus A\in\mathcal F$;
3. $\mathcal F$ è chiusa per unioni **numerabili**: $A_1,A_2,\ldots\in\mathcal F\Rightarrow\bigcup_{n=1}^\infty A_n\in\mathcal F$.

Gli elementi di $\mathcal F$ si dicono **insiemi misurabili**. La coppia $(X,\mathcal F)$ si chiama **spazio misurabile**.

**Conseguenze immediate.** $\emptyset=X\setminus X\in\mathcal F$ (per gli assiomi 1 e 2). Per le **leggi di De Morgan** (notebook 0.13), $\mathcal F$ è chiusa anche per **intersezioni numerabili**: $\bigcap_n A_n=X\setminus\bigcup_n(X\setminus A_n)$, e ogni $X\setminus A_n\in\mathcal F$ per l'assioma 2, la loro unione numerabile sta in $\mathcal F$ per l'assioma 3, e il complementare finale ancora per l'assioma 2. Dunque una σ-algebra è automaticamente chiusa anche per **differenze** ($A\setminus B=A\cap(X\setminus B)$) e differenze simmetriche.

### Confronto con la topologia: σ-algebra più forte, meno esempi "naturali"

Una σ-algebra è "più esigente" di una topologia: mentre una topologia tollera intersezioni infinite arbitrarie che falliscono (notebook 12.1, l'esempio $\bigcap_n(-1/n,1/n)$), una σ-algebra **richiede** che ogni unione o intersezione numerabile di misurabili resti misurabile — nessuna eccezione. Questo la rende una struttura più rigida, ma è proprio ciò che serve per la teoria della misura: **vogliamo** che $\{0\}=\bigcap_n(-1/n,1/n)$ sia misurabile (di misura zero, come vedremo nel notebook 13.2), non che sfugga alla struttura.

### Esempi fondamentali

- **σ-algebra banale**: $\mathcal F=\{\emptyset,X\}$. La più povera possibile (verifica immediata degli assiomi).
- **σ-algebra discreta**: $\mathcal F=\mathcal P(X)$ (ogni sottoinsieme misurabile). La più ricca possibile: banalmente chiusa per ogni operazione, essendo $\mathcal P(X)$ chiusa per qualunque unione, intersezione, complementare (stesso argomento del notebook 12.1 per la topologia discreta).
- **σ-algebra generata da una famiglia $\mathcal E$**: l'intersezione di **tutte** le σ-algebre che contengono $\mathcal E$ (esiste sempre, essendo l'intersezione di σ-algebre ancora una σ-algebra — dimostrazione identica, punto per punto, a quella dell'esercizio 3 del notebook 12.1 per le topologie). È la σ-algebra "più piccola" contenente $\mathcal E$, indicata $\sigma(\mathcal E)$.

### La σ-algebra di Borel

La **σ-algebra di Borel** su $\mathbb R$ (o più in generale su uno spazio topologico $X$), indicata $\mathcal B(\mathbb R)$, è $\sigma(\tau)$: la σ-algebra generata dagli aperti della topologia euclidea (notebook 12.1). Contiene tutti gli aperti, tutti i chiusi (complementari di aperti), tutte le unioni e intersezioni numerabili di questi — praticamente ogni insieme che si incontra in pratica. Un insieme in $\mathcal B(\mathbb R)$ si dice **boreliano**.

**Esempio concreto: $\{0\}$ è boreliano.** $\{0\}=\bigcap_n(-1/n,1/n)$: intersezione numerabile di aperti, quindi (per la chiusura di $\mathcal B(\mathbb R)$ rispetto a intersezioni numerabili, dedotta sopra) $\{0\}\in\mathcal B(\mathbb R)$, anche se $\{0\}$ non è esso stesso un aperto. Più in generale, **ogni** insieme numerabile (come $\mathbb Q$, notebook 10.5) è boreliano: unione numerabile dei singoletti $\{q\}$, ciascuno chiuso (quindi boreliano).

```python
import numpy as np

# Verifica computazionale: la sigma-algebra discreta su un insieme finito soddisfa gli assiomi
X = frozenset(range(4))  # X = {0,1,2,3}

def genera_sigma_algebra_discreta(X):
    from itertools import chain, combinations
    return [frozenset(c) for r in range(len(X)+1) for c in combinations(X, r)]

F = genera_sigma_algebra_discreta(X)
print(f"Numero di elementi della sigma-algebra discreta su |X|={len(X)}: {len(F)} (atteso 2^{len(X)}={2**len(X)})")

# Verifica assioma 2: chiusura per complementazione
chiusura_ok = all(frozenset(X - A) in F for A in F)
print("Chiusa per complementazione?", chiusura_ok)

# Verifica assioma 3 (unione finita, come approssimazione dell'unione numerabile su insieme finito)
import itertools
chiusura_unione = all(frozenset(A | B) in F for A, B in itertools.product(F, repeat=2))
print("Chiusa per unioni (finite, essendo X finito)?", chiusura_unione)
```

## Esercizio svolto e commentato

**Problema:** su $X=\{1,2,3,4\}$, trova la σ-algebra generata da $\mathcal E=\{\{1,2\}\}$, cioè $\sigma(\mathcal E)$.

**Ragionamento:**

**Passo 1 — partiamo dagli elementi obbligatori.** Deve contenere $X=\{1,2,3,4\}$ (assioma 1) e $\{1,2\}$ (per contenere $\mathcal E$).

**Passo 2 — applichiamo la chiusura per complementazione.** Il complementare di $\{1,2\}$ in $X$ è $\{3,4\}$: deve appartenere alla σ-algebra. Il complementare di $X$ è $\emptyset$: deve appartenere anch'esso (coerente con la conseguenza generale dedotta nella teoria).

**Passo 3 — verifichiamo se servono altri insiemi.** Controlliamo le unioni tra gli insiemi trovati finora: $\{1,2\}\cup\{3,4\}=X$ (già presente); $\{1,2\}\cup\emptyset=\{1,2\}$ (già presente); $\{3,4\}\cup\emptyset=\{3,4\}$ (già presente). Nessun nuovo insieme emerge. Controlliamo i complementari: il complementare di $\emptyset$ è $X$ (già presente).

**Passo 4 — verifica finale che l'insieme trovato sia effettivamente una σ-algebra.** $\mathcal F=\{\emptyset,\{1,2\},\{3,4\},X\}$: contiene $X$ (assioma 1) ✓; è chiusa per complementazione (ogni elemento ha il proprio complementare nell'insieme, verificato al passo 2-3) ✓; è chiusa per unioni (essendo un insieme finito con solo 4 elementi, basta verificare tutte le coppie, fatto al passo 3) ✓.

**Conclusione:** $\sigma(\{1,2\})=\{\emptyset,\{1,2\},\{3,4\},X\}$. **Osservazione metodologica:** generare una σ-algebra da un singolo insieme $A\subsetneq X$ produce sempre, come qui, la σ-algebra $\{\emptyset,A,X\setminus A,X\}$ (quattro elementi): questo è il caso base della costruzione generale, dove $A$ e il suo complementare "partizionano" $X$ in due blocchi, e ogni elemento della σ-algebra generata è un'unione di blocchi (qui: $\emptyset$, un blocco, l'altro blocco, o entrambi).

```python
# Verifica computazionale della sigma-algebra generata da {1,2} su X={1,2,3,4}
X = frozenset({1, 2, 3, 4})
A = frozenset({1, 2})

F = {frozenset(), A, frozenset(X - A), X}
print("Sigma-algebra generata:", [set(s) for s in F])

def verifica_sigma_algebra(X, F):
    if X not in F:
        return False, "manca X"
    for S in F:
        if frozenset(X - S) not in F:
            return False, f"manca il complementare di {set(S)}"
    for S1 in F:
        for S2 in F:
            if frozenset(S1 | S2) not in F:
                return False, f"manca l'unione di {set(S1)} e {set(S2)}"
    return True, "tutti gli assiomi verificati"

ok, messaggio = verifica_sigma_algebra(X, F)
print(f"E' una sigma-algebra? {ok} ({messaggio})")
```

## Riepilogo finale

- Una **σ-algebra** $\mathcal F$ su $X$ contiene $X$, è chiusa per complementazione e per unioni **numerabili** (e quindi, per De Morgan, anche per intersezioni numerabili): una struttura più esigente della topologia, pensata apposta per la teoria della misura.
- Gli elementi di una σ-algebra sono gli **insiemi misurabili**; $(X,\mathcal F)$ si chiama spazio misurabile.
- La **σ-algebra di Borel** $\mathcal B(\mathbb R)$, generata dagli aperti, contiene aperti, chiusi, e ogni insieme numerabile: la classe standard di insiemi "misurabili in pratica".
- $\sigma(\mathcal E)$, la σ-algebra generata da una famiglia $\mathcal E$, è la più piccola σ-algebra contenente $\mathcal E$: costruzione analoga, punto per punto, a quella della topologia generata da una base (notebook 12.2).
