## Perché ci interessa

Nel notebook 10.15 l'integrale di Riemann è stato costruito suddividendo il **dominio** in intervalli piccoli e sommando $f(x_i)\cdot(\text{larghezza})$. Questo funziona bene per funzioni "quasi continue", ma fallisce per la funzione di Dirichlet (notebook 13.3): ogni somma di Riemann, scegliendo punti razionali, dà $1$; scegliendo irrazionali, dà $0$ — non converge a un valore unico. L'idea di Lebesgue è ribaltare la prospettiva: invece di suddividere il **dominio**, suddividiamo il **codominio** (i valori assunti da $f$) e misuriamo **quanto dominio** produce ciascun valore. Questo cambio di visuale, reso possibile esattamente dalla teoria della misura (notebook 13.1-13.3), permette di integrare una classe di funzioni molto più ampia.

## Teoria, passo dopo passo

### Funzioni semplici

Una **funzione semplice** è una combinazione lineare finita di indicatrici di insiemi misurabili: $s(x)=\sum_{i=1}^n c_i\,\mathbb 1_{A_i}(x)$, con $c_i\ge0$ e $A_i\in\mathcal F$ disgiunti. Per una funzione semplice, l'integrale è **definito direttamente** come la somma pesata delle misure: $\int s\,d\mu=\sum_{i=1}^n c_i\,\mu(A_i)$ — l'analogo discreto di "valore per larghezza", ma ora "larghezza" è la misura di un insieme qualsiasi, non necessariamente un intervallo.

### Ogni funzione misurabile non negativa è limite crescente di funzioni semplici

**Teorema (costruzione, enunciato con schema di dimostrazione).** Se $f:X\to[0,+\infty]$ è misurabile, esiste una successione crescente di funzioni semplici $s_n\uparrow f$ (puntualmente, per ogni $x$).

**Costruzione esplicita.** Per ogni $n$, suddividiamo $[0,n)$ in $n\cdot2^n$ intervalli di lunghezza $2^{-n}$, e poniamo $s_n(x)=k/2^n$ se $f(x)\in[k/2^n,(k+1)/2^n)$ per $k=0,1,\ldots,n2^n-1$, e $s_n(x)=n$ se $f(x)\ge n$. Ogni $s_n$ è semplice (gli insiemi $\{x:f(x)\in[k/2^n,(k+1)/2^n)\}=f^{-1}([k/2^n,(k+1)/2^n))$ sono misurabili, essendo $f$ misurabile, notebook 13.3), e $s_n\le s_{n+1}\le f$ (raffinando la suddivisione, l'approssimazione migliora); inoltre $s_n(x)\to f(x)$ per ogni $x$ (per $n$ abbastanza grande da superare $f(x)$ e con la suddivisione $2^{-n}$ sempre più fine). Questa è, in sostanza, l'analogo per la misura della costruzione delle somme di Riemann inferiori (notebook 10.15), ma qui l'approssimazione avviene sui **valori** di $f$, non sul dominio.

### Definizione dell'integrale di Lebesgue

Per $f\ge0$ misurabile: $\int f\,d\mu=\sup\Big\{\int s\,d\mu:s\text{ semplice},0\le s\le f\Big\}$ (l'estremo superiore esiste sempre in $[0,+\infty]$, per completezza, notebook 10.9, essendo l'insieme non vuoto — contiene $s=0$). Per $f$ di segno qualsiasi, si scrive $f=f^+-f^-$ (parte positiva e negativa, $f^+=\max(f,0)$, $f^-=\max(-f,0)$, entrambe misurabili per il notebook 13.3) e si pone $\int f\,d\mu=\int f^+\,d\mu-\int f^-\,d\mu$, **quando almeno uno dei due termini è finito** (per evitare $\infty-\infty$). Se **entrambi** $\int f^+\,d\mu$ e $\int f^-\,d\mu$ sono finiti, $f$ si dice **integrabile** (Lebesgue).

### La funzione di Dirichlet è integrabile secondo Lebesgue

Riprendendo $D(x)=\mathbb 1_{\mathbb Q}(x)$ su $[0,1]$: $D$ è già una funzione semplice (un solo termine, $c_1=1$, $A_1=\mathbb Q\cap[0,1]$). Per definizione: $\int_{[0,1]}D\,d\lambda=1\cdot\lambda(\mathbb Q\cap[0,1])=1\cdot0=0$ (essendo $\mathbb Q$ numerabile, ha misura di Lebesgue nulla, notebook 13.2). **L'integrale di Lebesgue esiste e vale $0$**, mentre l'integrale di Riemann non esisteva affatto: la teoria della misura ha risolto esattamente il problema che ha motivato questo notebook.

### Confronto con Riemann: quando coincidono

**Teorema (enunciato).** Se $f$ è Riemann-integrabile su $[a,b]$, allora è anche Lebesgue-integrabile, e i due integrali coincidono. Il viceversa è falso (Dirichlet è un controesempio). Più precisamente, un teorema di Lebesgue (fuori portata qui) caratterizza le funzioni Riemann-integrabili come quelle **limitate e continue quasi ovunque** (concetto che preciseremo nel notebook 13.5): l'integrale di Lebesgue è dunque una **genuina estensione**, non un sostituto incompatibile, di quello di Riemann.

```python
import numpy as np

# Approssimiamo l'integrale di Lebesgue di f(x) = x^2 su [0,1] tramite funzioni semplici
# (approssimazione "orizzontale": suddividiamo il CODOMINIO, non il dominio)

def f(x):
    return x**2

def approssima_lebesgue(f, dominio, n_livelli):
    x_campione = np.linspace(*dominio, 100000)
    valori = f(x_campione)
    passo = 1.0 / n_livelli  # suddividiamo [0, max(f)] in n_livelli fasce
    integrale = 0.0
    for k in range(n_livelli):
        soglia_bassa, soglia_alta = k * passo, (k + 1) * passo
        # misura (approssimata) dell'insieme {x : f(x) in [soglia_bassa, soglia_alta)}
        indicatore = (valori >= soglia_bassa) & (valori < soglia_alta)
        misura_insieme = indicatore.sum() / len(x_campione) * (dominio[1] - dominio[0])
        integrale += soglia_bassa * misura_insieme  # funzione semplice: usa il valore minimo della fascia
    return integrale

for n in [5, 20, 100, 500]:
    stima = approssima_lebesgue(f, (0, 1), n)
    print(f"n_livelli={n}: integrale approssimato = {stima:.5f} (atteso: integrale di x^2 su [0,1] = 1/3 = {1/3:.5f})")
```

## Esercizio svolto e commentato

**Problema:** calcola $\int_{[0,1]}f\,d\lambda$ dove $f(x)=2$ se $x\in[0,1/2]$, $f(x)=5$ se $x\in(1/2,1]$ (una funzione a gradino), usando direttamente la definizione per funzioni semplici, e confronta con il calcolo tramite l'integrale di Riemann.

**Ragionamento:**

**Passo 1 — riconosciamo che $f$ è già una funzione semplice.** $f=2\cdot\mathbb 1_{[0,1/2]}+5\cdot\mathbb 1_{(1/2,1]}$, con $A_1=[0,1/2]$, $A_2=(1/2,1]$ disgiunti e misurabili (intervalli, quindi boreliani).

**Passo 2 — calcoliamo le misure di Lebesgue dei due pezzi.** $\lambda(A_1)=\lambda([0,1/2])=1/2$ (la misura di Lebesgue di un intervallo è la sua lunghezza, come richiesto dalla costruzione, notebook 13.2). $\lambda(A_2)=\lambda((1/2,1])=1/2$.

**Passo 3 — applichiamo la definizione.** $\int_{[0,1]}f\,d\lambda=2\cdot\lambda(A_1)+5\cdot\lambda(A_2)=2\cdot\frac12+5\cdot\frac12=1+2.5=3.5$.

**Passo 4 — confrontiamo con Riemann.** L'integrale di Riemann di una funzione a gradino, definito nel notebook 10.15 tramite somme superiori/inferiori, dà lo stesso risultato: $\int_0^{1/2}2\,dx+\int_{1/2}^1 5\,dx=2\cdot\frac12+5\cdot\frac12=3.5$. **Coincidono**, coerentemente con il teorema della teoria (ogni funzione Riemann-integrabile — e una funzione a gradino con un numero finito di salti lo è sempre — è anche Lebesgue-integrabile con lo stesso valore). La differenza tra i due approcci diventa visibile solo per funzioni patologiche come Dirichlet, non per funzioni "ragionevoli" come questa.

```python
import numpy as np

def f(x):
    return np.where(x <= 0.5, 2, 5)

# Calcolo diretto secondo la definizione di Lebesgue per funzioni semplici
misura_A1 = 0.5  # lambda([0, 1/2])
misura_A2 = 0.5  # lambda((1/2, 1])
integrale_lebesgue = 2 * misura_A1 + 5 * misura_A2
print(f"Integrale di Lebesgue (definizione diretta): {integrale_lebesgue}")

# Confronto con approssimazione numerica (alla Riemann, somme rettangolari)
x = np.linspace(0, 1, 1000000)
y = f(x)
integrale_riemann_approx = np.trapz(y, x)
print(f"Integrale di Riemann (approssimazione numerica): {integrale_riemann_approx:.5f}")
print(f"Coincidono? {abs(integrale_lebesgue - integrale_riemann_approx) < 1e-3}")
```

**1.** $\lambda([0,1])=1$, $\lambda((1,2])=1$. $\int f\,d\lambda=3\cdot1+7\cdot1=10$.

Dunque $\int(as+bt)\,d\mu=\sum_{i,j}(ac_i+bd_j)\mu(A_i\cap B_j)=a\sum_{i,j}c_i\mu(A_i\cap B_j)+b\sum_{i,j}d_j\mu(A_i\cap B_j)=a\int s\,d\mu+b\int t\,d\mu$.

**3.** Sia $N=\{x:f(x)\ne0\}$, con $\mu(N)=0$. Sia $s\le f$ una funzione semplice qualsiasi, $s=\sum_i c_i\mathbb 1_{A_i}$ con $c_i>0$: per ogni $i$ con $c_i>0$, deve essere $A_i\subseteq N$ (se $x\in A_i$ e $c_i>0$, allora $s(x)=c_i>0$, quindi $f(x)\ge s(x)>0$, cioè $x\in N$). Per monotonia della misura (notebook 13.2), $\mu(A_i)\le\mu(N)=0$, quindi $\mu(A_i)=0$. Dunque $\int s\,d\mu=\sum_i c_i\,\mu(A_i)=\sum_i c_i\cdot0=0$ per **ogni** $s\le f$ semplice. Per definizione, $\int f\,d\mu=\sup\{\int s\,d\mu:s\le f\}=\sup\{0\}=0$.

**4.** Definiamo $g(x)=1$ se $x\in\mathbb Q\cap[0,1]$, $g(x)=2$ se $x\in([0,1]\setminus\mathbb Q)\cap[0,1/2]$, $g(x)=3$ se $x\in([0,1]\setminus\mathbb Q)\cap(1/2,1]$: una funzione semplice a tre valori, i cui insiemi di livello sono ancora tutti boreliani ($\mathbb Q$ numerabile quindi boreliano, gli irrazionali intersecati con intervalli sono boreliani per complementazione). Ogni somma di Riemann, scegliendo punti razionali nella suddivisione, tenderebbe a dare valore vicino a $1$ ovunque; scegliendo irrazionali in $[0,1/2]$ e $(1/2,1]$ rispettivamente, si otterrebbero contributi da $2$ e $3$: le somme superiori e inferiori di Riemann non convergono allo stesso valore (essendo $g$ discontinua ovunque, con oscillazione locale sempre positiva, per densità di $\mathbb Q$ e del suo complementare), quindi $g$ non è Riemann-integrabile. Ma $g$ è Lebesgue-integrabile per definizione (funzione semplice): $\int g\,d\lambda=1\cdot\lambda(\mathbb Q\cap[0,1])+2\cdot\lambda(([0,1]\setminus\mathbb Q)\cap[0,1/2])+3\cdot\lambda(([0,1]\setminus\mathbb Q)\cap(1/2,1])=1\cdot0+2\cdot\frac12+3\cdot\frac12=0+1+1.5=2.5$ (usando che togliere un insieme di misura nulla, $\mathbb Q$, da un intervallo non ne cambia la misura).

## Riepilogo finale

- L'**integrale di Lebesgue** suddivide il **codominio** (invece del dominio, come Riemann): prima si definisce sulle funzioni semplici come somma pesata di misure, poi si estende per approssimazione crescente a tutte le funzioni misurabili non negative.
- Ogni funzione misurabile non negativa è limite crescente di funzioni semplici: la costruzione di Lebesgue funziona su qualunque funzione misurabile, indipendentemente da quanto sia "irregolare".
- La funzione di Dirichlet, non Riemann-integrabile, è Lebesgue-integrabile con integrale $0$: l'integrale di Lebesgue è una vera estensione di quello di Riemann, non un sostituto incompatibile.
- Una funzione nulla quasi ovunque ha sempre integrale nullo: la teoria della misura "ignora" automaticamente ciò che accade solo su insiemi di misura zero, un principio che approfondiremo nel prossimo notebook.
