# 5.18 — Integrali impropri

**Prerequisiti:** 5.16 Teorema fondamentale del calcolo, 4.2 Limiti di successioni, 3.12 Funzioni razionali.


## Perché ci interessa

La teoria dell'integrale di Riemann (notebook 5.15) richiede due condizioni: intervallo **limitato** e funzione **limitata**. Ma ha senso chiedersi "qual è l'area sotto $\frac1{x^2}$ da $1$ a $+\infty$?" (intervallo illimitato), o "qual è l'area sotto $\frac1{\sqrt x}$ da $0$ a $1$?" (funzione illimitata vicino a $0$). Gli **integrali impropri** estendono la teoria a questi casi, usando esattamente lo strumento che abbiamo costruito per le successioni: il limite.

## Teoria, passo dopo passo

### Integrali impropri su intervallo illimitato

Se $f$ è integrabile su $[a,T]$ per ogni $T>a$, definiamo:

$$\int_a^{+\infty}f(x)\,dx = \lim_{T\to+\infty}\int_a^Tf(x)\,dx$$

Se questo limite esiste finito, l'integrale improprio **converge** (esattamente il linguaggio delle serie, notebook 4.7 — non a caso: le somme parziali di una serie sono l'analogo discreto di $\int_a^Tf$, e la loro convergenza per $N\to\infty$ è l'analogo di questo limite per $T\to\infty$). Se il limite è infinito o non esiste, l'integrale **diverge**.

**Esempio guida**: $\int_1^{+\infty}\dfrac1{x^2}\,dx$. Calcoliamo prima l'integrale su $[1,T]$ (teorema fondamentale, notebook 5.16): $\int_1^T\dfrac1{x^2}\,dx=\left[-\dfrac1x\right]_1^T=-\dfrac1T+1$. Il limite per $T\to+\infty$: $-\dfrac1T\to0$ (notebook 4.2), quindi:

$$\int_1^{+\infty}\frac1{x^2}\,dx = 1$$

Un fatto sorprendente: una regione **illimitata** (che si estende all'infinito) ha area **finita**!

### Il confronto con $\dfrac1{x^p}$: un criterio generale

Con lo stesso procedimento si mostra che $\int_1^{+\infty}\dfrac1{x^p}\,dx$ converge se e solo se $p>1$ (diverge per $p\le1$, incluso il caso $p=1$: $\int_1^{+\infty}\dfrac1x\,dx$, l'analogo continuo della serie armonica divergente, notebook 4.9!). Questo criterio è l'analogo continuo di quello per le serie $\sum\frac1{n^p}$, e i due fenomeni sono profondamente collegati (un fatto noto come **criterio integrale** per le serie, che vedremo accennato nel notebook 5.19).

### Integrali impropri di funzioni illimitate

Se $f$ non è limitata vicino a un estremo (per esempio $b$), ma è integrabile su $[a,b-\varepsilon]$ per ogni $\varepsilon>0$ piccolo, definiamo:

$$\int_a^bf(x)\,dx = \lim_{\varepsilon\to0^+}\int_a^{b-\varepsilon}f(x)\,dx$$

**Esempio guida**: $\int_0^1\dfrac1{\sqrt x}\,dx$ ($f$ non limitata vicino a $0$). Calcoliamo su $[\varepsilon,1]$: $\int_\varepsilon^1x^{-1/2}\,dx=\big[2\sqrt x\big]_\varepsilon^1=2-2\sqrt\varepsilon$. Per $\varepsilon\to0^+$: $2\sqrt\varepsilon\to0$ (continuità della radice, notebook 5.6), quindi l'integrale converge a $2$.

### Un criterio del confronto, analogo a quello delle serie

Se $0\le f(x)\le g(x)$ per $x$ grande (o vicino al punto critico), e $\int g$ converge, allora anche $\int f$ converge (e viceversa per la divergenza) — esattamente il criterio del confronto per le serie (notebook 4.10), applicato ora agli integrali.

````python
import sympy as sp

x, T, epsilon = sp.symbols('x T epsilon', positive=True)

# Integrale improprio su intervallo illimitato: 1/x^2 da 1 a infinito
integrale_finito = sp.integrate(1/x**2, (x, 1, T))
print("Integrale su [1,T]:", integrale_finito)
limite = sp.limit(integrale_finito, T, sp.oo)
print("Limite per T->infinito:", limite)

# Confronto: 1/x da 1 a infinito (diverge, come la serie armonica)
integrale_1_su_x = sp.integrate(1/x, (x, 1, T))
print("\nIntegrale di 1/x su [1,T]:", integrale_1_su_x)
print("Limite per T->infinito:", sp.limit(integrale_1_su_x, T, sp.oo))

# Integrale improprio di funzione illimitata: 1/sqrt(x) da 0 a 1
integrale_radice = sp.integrate(1/sp.sqrt(x), (x, epsilon, 1))
print("\nIntegrale su [epsilon,1]:", integrale_radice)
limite_radice = sp.limit(integrale_radice, epsilon, 0, '+')
print("Limite per epsilon->0+:", limite_radice)

# Verifica diretta con SymPy (integra direttamente riconoscendo l'improprietà)
print("\nVerifica diretta:", sp.integrate(1/x**2, (x, 1, sp.oo)))
print("Verifica diretta:", sp.integrate(1/sp.sqrt(x), (x, 0, 1)))
````

## Esercizio svolto e commentato

**Problema:** studia la convergenza di $\displaystyle\int_0^{+\infty}e^{-x}\,dx$ (un integrale doppiamente "esteso": non serve qui, ma il metodo generale gestisce anche estremi infiniti su entrambi i lati spezzando in due pezzi).

**Ragionamento:**

**Passo 1 — impostiamo il limite** secondo la definizione:

$$\int_0^{+\infty}e^{-x}\,dx = \lim_{T\to+\infty}\int_0^Te^{-x}\,dx$$

**Passo 2 — calcoliamo l'integrale su $[0,T]$**, trovando prima una primitiva di $e^{-x}$: per la regola della catena (notebook 5.9), $\frac{d}{dx}(-e^{-x})=-(-e^{-x})=e^{-x}$, quindi $-e^{-x}$ è una primitiva.

$$\int_0^Te^{-x}\,dx = \big[-e^{-x}\big]_0^T = -e^{-T}-(-e^0) = 1-e^{-T}$$

**Passo 3 — calcoliamo il limite per $T\to+\infty$.** Sappiamo (notebook 1.11, 3.13) che $e^{-T}=\frac1{e^T}\to0$ per $T\to+\infty$ (l'esponenziale a denominatore cresce senza limite):

$$\int_0^{+\infty}e^{-x}\,dx = \lim_{T\to+\infty}(1-e^{-T}) = 1-0 = 1$$

**Conclusione**: l'integrale converge, e vale $1$.

**Interpretazione probabilistica** (un'anticipazione di argomenti che approfondirai nel Modulo 17): questo risultato è precisamente il motivo per cui $f(x)=e^{-x}$ (per $x\ge0$) può essere usata come **densità di probabilità** (la distribuzione esponenziale): l'area totale sotto la curva, su tutto il dominio, vale esattamente $1$ — la condizione di normalizzazione richiesta per ogni distribuzione di probabilità.

**Perché la convergenza di $\int_0^{+\infty}e^{-x}dx$ non è affatto scontata solo guardando che $e^{-x}\to0$:** come per le successioni (notebook 4.2) e le serie (notebook 4.9, l'esempio della serie armonica), il fatto che l'integrando tenda a $0$ non basta a garantire la convergenza dell'integrale improprio — serve che tenda a $0$ **abbastanza velocemente**. Qui $e^{-x}$ decade rapidissimamente (più veloce di ogni potenza $\frac1{x^p}$, coerentemente con quanto visto sul confronto tra esponenziali e polinomi, notebook 3.13), garantendo la convergenza; una funzione che decade troppo lentamente (come $\frac1x$, visto nella teoria) produce invece un integrale divergente.

````python
import sympy as sp

x, T = sp.symbols('x T', positive=True)
integrale_finito = sp.integrate(sp.exp(-x), (x, 0, T))
print("Integrale su [0,T]:", integrale_finito)

limite = sp.limit(integrale_finito, T, sp.oo)
print("Limite per T->infinito:", limite)

print("\nVerifica diretta:", sp.integrate(sp.exp(-x), (x, 0, sp.oo)))
````

## Riepilogo finale

- Integrali su intervallo illimitato: $\int_a^{+\infty}f=\lim_{T\to\infty}\int_a^Tf$; convergono o divergono a seconda che il limite sia finito o meno.
- Integrali di funzioni illimitate: si sostituisce l'estremo problematico con $b-\varepsilon$ (o $a+\varepsilon$) e si passa al limite per $\varepsilon\to0^+$.
- $\int_1^{+\infty}\frac1{x^p}dx$ converge se e solo se $p>1$: lo stesso criterio (con la stessa soglia) delle serie $\sum\frac1{n^p}$, un parallelo profondo tra le due teorie.
- Come per le serie, non basta che l'integrando tenda a $0$: serve che lo faccia abbastanza velocemente, e il criterio del confronto (analogo al notebook 4.10) aiuta a stabilirlo senza calcoli espliciti.
