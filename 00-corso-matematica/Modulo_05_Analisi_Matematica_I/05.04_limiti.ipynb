# 5.4 — Limiti

**Prerequisiti:** 5.3 Successioni (rigorose), 3.3 Dominio, 0.14 Intervalli.


## Perché ci interessa

Finora il limite era definito solo per successioni: $a_n\to L$ quando $n\to\infty$ (un indice discreto). Ma vogliamo studiare anche il comportamento di una funzione $f(x)$ quando $x$ si avvicina a un punto $x_0$ **con continuità** (non a scatti come $n\in\mathbb{N}$). Questo notebook definisce il **limite di una funzione**, con la stessa struttura logica (ε-δ invece di ε-N) usata per le successioni, aprendo la strada a continuità e derivata.

## Teoria, passo dopo passo

### La definizione ε-δ

Sia $f$ definita in un intorno di $x_0$ (tranne al più in $x_0$ stesso — il valore di $f$ **in** $x_0$, se esiste, non conta per il limite). Diciamo che:

$$\lim_{x\to x_0} f(x) = L$$

se, per ogni $\varepsilon>0$, esiste $\delta>0$ (che dipende da $\varepsilon$) tale che:

$$0<|x-x_0|<\delta \quad\Rightarrow\quad |f(x)-L|<\varepsilon$$

Il parallelo con le successioni (notebook 4.2, 5.3) è diretto: $\varepsilon$ misura ancora la tolleranza sul valore di $f$; $\delta$ gioca il ruolo che aveva $N$, ma misura una **distanza** (quanto vicino deve stare $x$ a $x_0$), non un indice. La condizione $0<|x-x_0|$ esclude $x=x_0$ stesso: il limite riguarda il comportamento **vicino** a $x_0$, non necessariamente il valore preciso in quel punto.

### Verifica dell'esempio più semplice: $\lim_{x\to2}(3x+1)=7$

Fissiamo $\varepsilon>0$. Vogliamo $|(3x+1)-7|<\varepsilon$, cioè $|3x-6|<\varepsilon$, cioè $3|x-2|<\varepsilon$ (raccogliendo $3$, notebook 0.5), cioè $|x-2|<\frac\varepsilon3$. Basta scegliere $\delta=\frac\varepsilon3$: allora $0<|x-2|<\delta \Rightarrow |x-2|<\frac\varepsilon3 \Rightarrow |(3x+1)-7|<\varepsilon$. Esattamente lo schema del notebook 5.3, applicato ora a una distanza $\delta$ invece che a un indice $N$.

### Limite e continuità (anteprima)

Se $f(x_0)$ è definito **e** coincide con $\lim_{x\to x_0}f(x)$, diremo che $f$ è **continua** in $x_0$ (formalizzato nel notebook 5.6). Per ora osserviamo solo che il limite può esistere anche se $f(x_0)$ non è definito (come nella funzione razionale con un "buco", notebook 3.12), o può esistere ma essere diverso da $f(x_0)$.

### Limiti destro e sinistro

A volte $f(x)$ si comporta diversamente avvicinandosi a $x_0$ da destra o da sinistra. Definiamo:

$$\lim_{x\to x_0^+}f(x)=L \quad (\text{limite destro}), \qquad \lim_{x\to x_0^-}f(x)=L \quad (\text{limite sinistro})$$

restringendo la condizione a $x_0<x<x_0+\delta$ (destro) o $x_0-\delta<x<x_0$ (sinistro). Il limite "bilatero" $\lim_{x\to x_0}f(x)$ esiste **se e solo se** esistono entrambi i limiti unilaterali **e coincidono**. Un esempio dove differiscono: la funzione a tratti $f(x)=|x|/x$ per $x\ne0$ (notebook 0.8, 1.14): $\lim_{x\to0^+}f(x)=1$ ma $\lim_{x\to0^-}f(x)=-1$, quindi $\lim_{x\to0}f(x)$ **non esiste**.

### Limiti all'infinito e limiti infiniti

Estendiamo la definizione a due casi analoghi a quelli già visti per le successioni (notebook 4.2): $\lim_{x\to+\infty}f(x)=L$ (il comportamento di $f$ per $x$ grande, analogo diretto al limite di successione, con $\delta$ sostituito da un'ampiezza sull'asse $x$); e $\lim_{x\to x_0}f(x)=+\infty$ (la funzione cresce senza limite avvicinandosi a $x_0$ — il comportamento già osservato informalmente per gli asintoti verticali, notebook 3.12).

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica della definizione epsilon-delta per lim (3x+1) = 7 quando x->2
def trova_delta(epsilon):
    return epsilon / 3

for eps in [0.3, 0.03, 0.003]:
    delta = trova_delta(eps)
    print(f"epsilon={eps}: delta={delta}")
    # verifica su un punto dentro l'intervallo (2-delta, 2+delta)
    x_test = 2 + delta/2
    valore = 3*x_test + 1
    print(f"  x={x_test}: f(x)={valore}, |f(x)-7|={abs(valore-7):.5f} < {eps}? {abs(valore-7) < eps}")

# Limite destro e sinistro diversi: f(x) = |x|/x
f = sp.Abs(x) / x
limite_destro = sp.limit(f, x, 0, dir='+')
limite_sinistro = sp.limit(f, x, 0, dir='-')
print(f"\nlim x->0+ |x|/x = {limite_destro}")
print(f"lim x->0- |x|/x = {limite_sinistro}")
print("Il limite bilatero esiste?", limite_destro == limite_sinistro)
````

## Esercizio svolto e commentato

**Problema:** dimostra con la definizione ε-δ che $\lim_{x\to3}x^2=9$.

**Ragionamento:**

**Passo 1 — impostiamo la disequazione target:** vogliamo $|x^2-9|<\varepsilon$ per $x$ vicino a $3$.

**Passo 2 — scomponiamo** (differenza di quadrati, notebook 1.2): $|x^2-9|=|x-3||x+3|$.

**Passo 3 — la difficoltà (rispetto all'esempio lineare della teoria): il fattore $|x+3|$ non è costante.** Serve limitarlo, sapendo che $x$ è vicino a $3$. **Restringiamo preventivamente** $\delta\le1$: allora $|x-3|<1$ implica $2<x<4$, quindi $5<x+3<7$, cioè $|x+3|<7$.

**Passo 4 — con questa restrizione**, $|x^2-9|=|x-3||x+3|<7|x-3|$. Vogliamo che questo sia $<\varepsilon$: basta $|x-3|<\frac\varepsilon7$.

**Passo 5 — combiniamo le due condizioni su $\delta$** (quella di restrizione, $\delta\le1$, e quella dell'obiettivo, $\delta\le\frac\varepsilon7$): scegliamo

$$\delta = \min\left(1,\frac\varepsilon7\right)$$

**Verifica finale**: se $0<|x-3|<\delta$, allora $|x-3|<1$ (quindi $|x+3|<7$, dal passo 3) **e** $|x-3|<\frac\varepsilon7$ (dal passo 5, essendo $\delta$ il minimo di entrambe le soglie). Quindi $|x^2-9|=|x-3||x+3|<\frac\varepsilon7\times7=\varepsilon$. ∎

**Perché la tecnica del "$\delta=\min(1,\dots)$" è necessaria qui e non nell'esempio lineare della teoria:** con $f(x)=3x+1$, il coefficiente di $|x-3|$ era **costante** ($3$), quindi bastava dividere $\varepsilon$ per quella costante. Con $f(x)=x^2$, il "coefficiente" $|x+3|$ **dipende da $x$**: dobbiamo prima **limitarlo** con una restrizione preliminare su $\delta$ (qui, $\delta\le1$), per poi trattarlo come se fosse (al massimo) una costante nota. Questa tecnica del "doppio vincolo su $\delta$" è standard per funzioni non lineari, e la ritroveremo spesso.

````python
import sympy as sp

x = sp.symbols('x', real=True)

def delta_per_epsilon(epsilon):
    return min(1, epsilon/7)

for eps in [0.7, 0.07]:
    delta = delta_per_epsilon(eps)
    print(f"epsilon={eps}: delta={delta}")
    x_test = 3 + delta/2
    valore = x_test**2
    print(f"  x={x_test}: f(x)={valore}, |f(x)-9|={abs(valore-9):.5f} < {eps}? {abs(valore-9) < eps}")

# Verifica diretta con SymPy
limite = sp.limit(x**2, x, 3)
print("\nLimite calcolato da SymPy:", limite)
````

## Riepilogo finale

- $\lim_{x\to x_0}f(x)=L$ significa: per ogni $\varepsilon>0$ esiste $\delta>0$ tale che $0<|x-x_0|<\delta \Rightarrow |f(x)-L|<\varepsilon$ — la stessa struttura logica del limite di successioni, con $\delta$ al posto di $N$.
- Il limite non richiede che $f$ sia definita in $x_0$, solo vicino a esso; se lo è e coincide con il limite, la funzione è continua in quel punto (notebook 5.6).
- Il limite bilatero esiste solo se i limiti destro e sinistro esistono e coincidono.
- Per funzioni non lineari, il calcolo di $\delta$ spesso richiede una restrizione preliminare (es. $\delta\le1$) per limitare fattori non costanti prima di ottenere la scelta finale di $\delta$.
