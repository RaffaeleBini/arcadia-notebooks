# 5.10 — Teorema di Rolle

**Prerequisiti:** 5.9 Teoremi fondamentali della derivazione, 5.7 Teoremi fondamentali sulle funzioni continue.


## Perché ci interessa

Se una funzione parte e finisce allo stesso livello ($f(a)=f(b)$), l'intuizione geometrica suggerisce che, da qualche parte nel mezzo, il grafico deve "fare un giro" — salire e poi ridiscendere, o viceversa — e in quel punto la tangente deve essere orizzontale. Il **teorema di Rolle** rende rigorosa questa intuizione, ed è il primo mattone di una catena di teoremi (Rolle → Lagrange → Cauchy) che collegano la derivata al comportamento globale di una funzione.

## Teoria, passo dopo passo

### Enunciato

> Sia $f$ continua su $[a,b]$ e derivabile su $(a,b)$. Se $f(a)=f(b)$, allora esiste $c\in(a,b)$ tale che $f'(c)=0$.

Tre ipotesi, tutte necessarie: continuità su $[a,b]$ (chiuso), derivabilità su $(a,b)$ (aperto — non serve derivabilità agli estremi), e $f(a)=f(b)$.

### Dimostrazione

Per il teorema di Weierstrass (notebook 5.7), $f$ ammette massimo e minimo su $[a,b]$ (essendo continua su un chiuso e limitato).

**Caso 1**: se massimo e minimo coincidono, $f$ è **costante** su $[a,b]$ (l'immagine è un solo punto), quindi $f'(x)=0$ per ogni $x\in(a,b)$: qualunque $c$ funziona.

**Caso 2**: se massimo e minimo sono diversi, almeno uno dei due (chiamiamolo $M$, raggiunto in un punto $c$) è **diverso** da $f(a)=f(b)$ (altrimenti entrambi coinciderebbero con quel valore comune, e saremmo nel caso 1). Questo punto $c$ non può essere né $a$ né $b$ (essendo il valore lì diverso), quindi $c\in(a,b)$: un punto **interno**.

In un punto interno di massimo (o minimo) dove $f$ è derivabile, vale il **teorema di Fermat** (che enunciamo qui senza dimostrarlo per esteso, ma l'idea è diretta): se $c$ è un massimo interno, il rapporto incrementale $\frac{f(c+h)-f(c)}h$ è $\le0$ per $h>0$ piccolo (perché $f(c+h)\le f(c)$, essendo $c$ un massimo) e $\ge0$ per $h<0$ (stesso motivo, ma il denominatore cambia segno). Passando al limite $h\to0$ da entrambi i lati, essendo $f$ derivabile in $c$ (i due limiti unilaterali devono coincidere, notebook 5.4), l'unico valore compatibile con "$\le0$ da un lato e $\ge0$ dall'altro" è $f'(c)=0$. ∎

### Perché ogni ipotesi è necessaria

- **Senza continuità su $[a,b]$**: $f(x)=x$ su $[0,1)$ con $f(1)$ ridefinito artificialmente a $0$ (per avere $f(0)=f(1)=0$) — discontinua in $1$, e non esiste nessun punto con derivata $0$ (la funzione è sempre crescente con $f'=1$).
- **Senza derivabilità su $(a,b)$**: $f(x)=|x|$ su $[-1,1]$, con $f(-1)=f(1)=1$: continua, ma non derivabile in $0$ (notebook 5.8) — e infatti $f'(x)=\pm1$ ovunque tranne in $0$, mai $0$.
- **Senza $f(a)=f(b)$**: una retta con pendenza costante diversa da zero, come $f(x)=x$ su $[0,1]$: non c'è nessun punto con tangente orizzontale.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica del teorema di Rolle su f(x) = x^2 - 4x + 3 su [1,3] (f(1)=f(3)=0)
f = x**2 - 4*x + 3
print(f"f(1) = {f.subs(x,1)}, f(3) = {f.subs(x,3)}")

f_prime = sp.diff(f, x)
c_soluzioni = sp.solve(sp.Eq(f_prime, 0), x)
print(f"f'(x) = {f_prime}, si annulla in x = {c_soluzioni}")
print(f"Il punto è in (1,3)? {[1 < c < 3 for c in c_soluzioni]}")

# Grafico per visualizzare
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(1, 3, 100)
y_vals = x_vals**2 - 4*x_vals + 3
fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(x_vals, y_vals)
ax.axhline(y=0, color='gray', linewidth=0.5)
ax.plot([1, 3], [0, 0], 'ro', label='f(1)=f(3)=0')
c_val = float(c_soluzioni[0])
ax.plot(c_val, f.subs(x, c_val), 'g^', markersize=10, label=f'c={c_val} (tangente orizzontale)')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("Teorema di Rolle: esiste c con f'(c)=0")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** verifica le ipotesi del teorema di Rolle per $f(x)=\sin x$ su $[0,\pi]$, e trova esplicitamente il punto $c$.

**Ragionamento:**

**Passo 1 — verifichiamo la continuità su $[0,\pi]$.** $\sin x$ è continua su tutto $\mathbb{R}$ (notebook 5.6), quindi in particolare su $[0,\pi]$. ✓

**Passo 2 — verifichiamo la derivabilità su $(0,\pi)$.** $\sin x$ è derivabile ovunque, con $(\sin x)'=\cos x$ (catalogo del notebook 5.9). ✓

**Passo 3 — verifichiamo $f(a)=f(b)$.** $f(0)=\sin(0)=0$, $f(\pi)=\sin(\pi)=0$ (valore notevole, notebook 2.9): $f(0)=f(\pi)=0$. ✓

**Passo 4 — tutte le ipotesi sono soddisfatte: il teorema garantisce l'esistenza di $c\in(0,\pi)$ con $f'(c)=0$.** Troviamolo esplicitamente: $f'(c)=\cos c=0$. Dalla teoria del notebook 2.11 (equazioni trigonometriche elementari), $\cos c=0$ per $c=\frac\pi2+k\pi$. L'unico valore in $(0,\pi)$ è $c=\frac\pi2$.

**Verifica geometrica**: $\frac\pi2$ è esattamente il punto medio dell'intervallo $[0,\pi]$, dove $\sin x$ raggiunge il suo **massimo** (valore $1$, notebook 2.9) — coerente con la dimostrazione del teorema, che collegava il punto $c$ a un massimo o minimo interno.

**Perché in questo esempio il punto $c$ coincide esattamente con il punto di massimo, e se questo capiti sempre:** nella dimostrazione del teorema (teoria), $c$ è stato costruito **precisamente** come un punto di massimo o minimo interno (quando la funzione non è costante). Non è quindi un caso che qui $c=\frac\pi2$ sia il massimo: è esattamente lo schema previsto dalla dimostrazione. In generale, per funzioni più complesse (con più massimi/minimi locali, che vedremo nel Modulo 5.13), il teorema garantisce **almeno un** punto con $f'(c)=0$, ma potrebbero essere più di uno, e non necessariamente il massimo o minimo "globale" — solo uno **locale**, un concetto che approfondiremo studiando le funzioni in dettaglio.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = sp.sin(x)

print(f"f(0) = {f.subs(x, 0)}, f(pi) = {f.subs(x, sp.pi)}")

f_prime = sp.cos(x)
c_soluzioni = sp.solveset(sp.Eq(f_prime, 0), x, domain=sp.Interval.open(0, sp.pi))
print(f"Punti con f'(c)=0 in (0,pi): {c_soluzioni}")
````

## Riepilogo finale

- **Teorema di Rolle**: $f$ continua su $[a,b]$, derivabile su $(a,b)$, con $f(a)=f(b)$, garantisce l'esistenza di $c\in(a,b)$ con $f'(c)=0$.
- La dimostrazione combina il teorema di Weierstrass (esistenza di massimo/minimo, notebook 5.7) con il teorema di Fermat (derivata nulla in un punto di estremo interno).
- Ciascuna delle tre ipotesi è essenziale: senza continuità, senza derivabilità, o senza $f(a)=f(b)$, il teorema può fallire.
- Un'applicazione potente: dimostrare che un'equazione ha **al più** una soluzione, mostrando che la derivata non si annulla mai (altrimenti, due soluzioni implicherebbero, per Rolle, un punto con derivata nulla).
