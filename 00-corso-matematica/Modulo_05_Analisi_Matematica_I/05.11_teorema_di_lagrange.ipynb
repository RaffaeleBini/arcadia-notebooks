# 5.11 — Teorema di Lagrange

**Prerequisiti:** 5.10 Teorema di Rolle, 2.2 Retta.


## Perché ci interessa

Il teorema di Rolle (notebook 5.10) richiede $f(a)=f(b)$: un caso speciale. Ma cosa succede in generale, quando $f(a)\ne f(b)$? Il **teorema di Lagrange** (o del valor medio) risponde: esiste sempre un punto dove la tangente è **parallela** alla retta secante che unisce $(a,f(a))$ e $(b,f(b))$. È probabilmente il teorema più usato di tutta l'analisi, e da esso deriveremo (nei prossimi notebook) il collegamento rigoroso tra segno della derivata e monotonia.

## Teoria, passo dopo passo

### Enunciato

> Sia $f$ continua su $[a,b]$ e derivabile su $(a,b)$. Allora esiste $c\in(a,b)$ tale che:
> $$f'(c) = \frac{f(b)-f(a)}{b-a}$$

Il lato destro è esattamente il coefficiente angolare della retta secante per $(a,f(a))$ e $(b,f(b))$ (notebook 2.2): il teorema dice che la tangente in $c$ ha **la stessa pendenza** di questa secante.

### Dimostrazione: ricondursi a Rolle

L'idea è costruire una funzione ausiliaria che soddisfi le ipotesi del teorema di Rolle (notebook 5.10), "sottraendo" a $f$ la retta secante stessa. Definiamo:

$$g(x) = f(x) - f(a) - \frac{f(b)-f(a)}{b-a}(x-a)$$

($g$ misura quanto $f$ si discosta dalla retta secante in ogni punto). Verifichiamo le ipotesi di Rolle per $g$:

- $g$ è continua su $[a,b]$ e derivabile su $(a,b)$ (differenza di $f$, continua/derivabile per ipotesi, e di una funzione lineare, sempre continua e derivabile, notebook 2.2, 5.9).
- $g(a) = f(a)-f(a)-\frac{f(b)-f(a)}{b-a}(a-a) = 0$.
- $g(b) = f(b)-f(a)-\frac{f(b)-f(a)}{b-a}(b-a) = f(b)-f(a)-(f(b)-f(a)) = 0$.

Quindi $g(a)=g(b)=0$: per il teorema di Rolle, esiste $c\in(a,b)$ con $g'(c)=0$. Calcoliamo $g'(x)$ (regola della somma, notebook 5.9):

$$g'(x) = f'(x) - \frac{f(b)-f(a)}{b-a}$$

Da $g'(c)=0$:

$$f'(c) = \frac{f(b)-f(a)}{b-a}$$

esattamente la tesi. ∎ Il teorema di Rolle è quindi un **caso particolare** del teorema di Lagrange (quando $f(a)=f(b)$, il lato destro è $0$): ma la dimostrazione procede nella direzione opposta, usando Rolle come strumento per dimostrare Lagrange.

### Interpretazione fisica: la velocità media

Se $f(t)$ rappresenta la posizione di un oggetto al tempo $t$, $\frac{f(b)-f(a)}{b-a}$ è la **velocità media** nell'intervallo $[a,b]$ (spazio percorso diviso tempo impiegato). Il teorema di Lagrange dice che, in **almeno un istante** $c$, la **velocità istantanea** $f'(c)$ (notebook 5.8) deve coincidere esattamente con la velocità media — un fatto intuitivo (se percorri $100$ km in $2$ ore, in qualche istante devi aver viaggiato esattamente a $50$ km/h, anche se hai accelerato e rallentato lungo il tragitto) ora dimostrato con rigore.

### Conseguenza immediata: derivata nulla ⟹ funzione costante

> Se $f'(x)=0$ per ogni $x$ in un intervallo, allora $f$ è **costante** su quell'intervallo.

**Dimostrazione**: presi due punti qualsiasi $a<b$ nell'intervallo, per Lagrange esiste $c\in(a,b)$ con $f(b)-f(a)=f'(c)(b-a)=0\times(b-a)=0$, quindi $f(a)=f(b)$. Essendo $a,b$ arbitrari, $f$ assume lo stesso valore ovunque: è costante. Questo risultato, apparentemente ovvio, sarà cruciale nel notebook 5.16 per giustificare che due primitive della stessa funzione differiscono solo per una costante.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica del teorema di Lagrange su f(x) = x^3 su [0,2]
f = x**3
a_val, b_val = 0, 2
pendenza_secante = (f.subs(x, b_val) - f.subs(x, a_val)) / (b_val - a_val)
print(f"Pendenza della secante: {pendenza_secante}")

f_prime = sp.diff(f, x)
c_soluzioni = sp.solve(sp.Eq(f_prime, pendenza_secante), x)
print(f"f'(x) = {pendenza_secante} risolta: c = {c_soluzioni}")
c_valido = [c for c in c_soluzioni if a_val < c < b_val]
print(f"Punto valido in (0,2): {c_valido}")

# Verifica del corollario: f'(x)=0 ovunque implica f costante
g = sp.Function('g')
x_sym = sp.symbols('x')
espressione_costante = sp.diff(sp.Rational(5), x_sym)  # derivata di una costante
print(f"\nDerivata della funzione costante f(x)=5: {espressione_costante}")

import matplotlib.pyplot as plt
import numpy as np
x_vals = np.linspace(0, 2, 100)
fig, ax = plt.subplots(figsize=(7,5))
ax.plot(x_vals, x_vals**3, label='f(x)=x³')
secante = pendenza_secante * (x_vals - a_val) + f.subs(x, a_val)
ax.plot(x_vals, [float(s) for s in secante], '--', label='secante')
c_val = float(c_valido[0])
tangente = f_prime.subs(x, c_val) * (x_vals - c_val) + f.subs(x, c_val)
ax.plot(x_vals, [float(t) for t in tangente], ':', label=f'tangente in c={c_val:.3f}')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("Teorema di Lagrange: tangente parallela alla secante")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** usa il teorema di Lagrange per dimostrare che, per ogni $x>0$, vale $\dfrac{x}{1+x}<\ln(1+x)<x$.

**Ragionamento:**

**Passo 1 — applichiamo Lagrange a $f(t)=\ln(1+t)$ sull'intervallo $[0,x]$** (con $x>0$ fissato). $f$ è continua e derivabile per $t>-1$ (dominio del logaritmo, notebook 1.12, 3.14), quindi in particolare su $[0,x]$.

**Passo 2 — calcoliamo il rapporto della secante:**

$$\frac{f(x)-f(0)}{x-0} = \frac{\ln(1+x)-\ln1}x = \frac{\ln(1+x)}x$$

**Passo 3 — per il teorema, esiste $c\in(0,x)$ tale che questo rapporto coincide con $f'(c)$.** Calcoliamo $f'(t)=\dfrac1{1+t}$ (regola della catena, notebook 5.9, con derivata del logaritmo dal catalogo):

$$\frac{\ln(1+x)}x = \frac1{1+c}$$

**Passo 4 — usiamo che $0<c<x$ per limitare $\frac1{1+c}$.** Essendo $\frac1{1+t}$ una funzione **decrescente** per $t>-1$ (il denominatore cresce, notebook 1.9): $c<x \Rightarrow 1+c<1+x \Rightarrow \frac1{1+c}>\frac1{1+x}$; e $c>0 \Rightarrow 1+c>1 \Rightarrow \frac1{1+c}<1$. Quindi:

$$\frac1{1+x} < \frac1{1+c} < 1$$

**Passo 5 — sostituiamo** $\frac1{1+c}=\frac{\ln(1+x)}x$ (dal passo 3):

$$\frac1{1+x} < \frac{\ln(1+x)}x < 1$$

Moltiplicando per $x>0$ (verso invariato, notebook 1.8):

$$\frac{x}{1+x} < \ln(1+x) < x$$

esattamente la disuguaglianza richiesta. ∎

**Perché il teorema di Lagrange è uno strumento così efficace per dimostrare disuguaglianze come questa:** invece di manipolare direttamente $\ln(1+x)$ (difficile da confrontare algebricamente con $x$), il teorema lo **sostituisce** con $f'(c)=\frac1{1+c}$ per un opportuno $c$ intermedio — un'espressione molto più facile da limitare (essendo una funzione monotona nota). Questa strategia (applicare Lagrange, poi limitare $f'(c)$ usando $a<c<b$) è uno degli usi più comuni del teorema in tutta l'analisi.

````python
import sympy as sp

x = sp.symbols('x', positive=True)

# Verifica numerica della disuguaglianza per alcuni valori
for x_val in [0.5, 1, 2, 5]:
    sinistra = x_val / (1 + x_val)
    centro = float(sp.log(1 + x_val))
    destra = x_val
    print(f"x={x_val}: {sinistra:.4f} < {centro:.4f} < {destra}?  {sinistra < centro < destra}")
````

## Riepilogo finale

- **Teorema di Lagrange**: $f$ continua su $[a,b]$, derivabile su $(a,b)$, garantisce $c\in(a,b)$ con $f'(c)=\dfrac{f(b)-f(a)}{b-a}$: la tangente in $c$ è parallela alla secante.
- Si dimostra riconducendosi al teorema di Rolle (notebook 5.10), tramite la funzione ausiliaria $g(x)=f(x)-$ (equazione della secante).
- **Corollario fondamentale**: $f'=0$ ovunque su un intervallo $\Rightarrow$ $f$ costante; più in generale, $f'=g'$ ovunque $\Rightarrow$ $f-g$ costante.
- Applicazione tipica: dimostrare disuguaglianze, sostituendo un'espressione complicata con $f'(c)$ e limitando quest'ultima usando $a<c<b$.
