## Teoria, passo dopo passo

### Il segno della derivata prima determina la monotonia

Questo è il collegamento cruciale, conseguenza diretta del teorema di Lagrange (notebook 5.11):

> Se $f'(x)>0$ per ogni $x$ in un intervallo, $f$ è strettamente crescente lì. Se $f'(x)<0$, strettamente decrescente.

**Dimostrazione** (caso crescente): presi $x_1<x_2$ nell'intervallo, per Lagrange esiste $c\in(x_1,x_2)$ con $f(x_2)-f(x_1)=f'(c)(x_2-x_1)$. Essendo $f'(c)>0$ e $x_2-x_1>0$, il prodotto è positivo: $f(x_2)>f(x_1)$, la definizione di crescita stretta (notebook 3.9). ∎

Questo trasforma lo studio della monotonia (prima solo intuitivo, notebook 3.9) in un problema di **studio del segno** di $f'(x)$ — esattamente le tecniche del notebook 1.9.

### I punti stazionari e la loro classificazione

Un punto dove $f'(x_0)=0$ si chiama **punto stazionario** (o critico). In un punto stazionario, studiando il segno di $f'$ **prima e dopo** $x_0$, si classifica:

- Se $f'$ passa da $+$ a $-$: $x_0$ è un **massimo locale**.
- Se $f'$ passa da $-$ a $+$: $x_0$ è un **minimo locale**.
- Se il segno non cambia: $x_0$ è un **flesso a tangente orizzontale** (né massimo né minimo).

### La derivata seconda e la concavità

La **derivata seconda** $f''(x)$ (la derivata della derivata) misura come cambia la pendenza: se $f''(x)>0$, la pendenza $f'(x)$ è crescente, cioè il grafico "curva verso l'alto" (**concavità verso l'alto**, come $x^2$, notebook 2.5); se $f''(x)<0$, **concavità verso il basso**. Un punto dove la concavità cambia si chiama **punto di flesso**.

**Test della derivata seconda** per i punti stazionari: se $f'(x_0)=0$ e $f''(x_0)>0$, $x_0$ è un minimo locale (il grafico è concavo verso l'alto lì, come il fondo di una parabola); se $f''(x_0)<0$, un massimo locale.

### Asintoti (ripasso sistematico)

Riprendiamo dal notebook 3.12: **verticali** dove il denominatore si annulla (e il numeratore no); **orizzontali** $y=L$ se $\lim_{x\to\pm\infty}f(x)=L$ (finito); **obliqui** $y=mx+q$ se $f(x)-mx-q\to0$, con $m=\lim_{x\to\infty}\frac{f(x)}x$ e $q=\lim_{x\to\infty}(f(x)-mx)$ — ora calcolabili rigorosamente con gli strumenti del notebook 5.4.

### Il procedimento completo

1. **Dominio** (notebook 3.3).
2. **Simmetrie** (pari/dispari, notebook 3.8) e intersezioni con gli assi.
3. **Limiti agli estremi del dominio** e **asintoti**.
4. **Segno di $f'$**: intervalli di crescita/decrescita, punti stazionari.
5. **Segno di $f''$**: concavità, punti di flesso.
6. **Sintesi**: disegno del grafico, coerente con tutte le informazioni raccolte.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Studio completo di f(x) = x^3 - 3x
f = x**3 - 3*x

print("=== Dominio ===")
print("R (polinomio)")

print("\n=== Simmetria ===")
print("f(-x) =", sp.expand(f.subs(x, -x)), " -> dispari" if sp.expand(f.subs(x,-x)) == -f else "")

print("\n=== Derivata prima e punti stazionari ===")
f1 = sp.diff(f, x)
print("f'(x) =", f1)
stazionari = sp.solve(sp.Eq(f1, 0), x)
print("Punti stazionari:", stazionari)

print("\n=== Derivata seconda (classificazione) ===")
f2 = sp.diff(f, x, 2)
print("f''(x) =", f2)
for p in stazionari:
    valore_f2 = f2.subs(x, p)
    tipo = "minimo locale" if valore_f2 > 0 else "massimo locale" if valore_f2 < 0 else "da approfondire"
    print(f"  in x={p}: f''={valore_f2} -> {tipo}")

print("\n=== Punti di flesso (f''=0) ===")
flessi = sp.solve(sp.Eq(f2, 0), x)
print("Candidati flesso:", flessi)

# Grafico riassuntivo
x_vals = np.linspace(-3, 3, 200)
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x_vals, x_vals**3 - 3*x_vals)
ax.axhline(y=0, color='gray', linewidth=0.5)
for p in stazionari:
    ax.plot(float(p), float(f.subs(x, p)), 'ro')
ax.grid(True, alpha=0.3)
ax.set_title("f(x) = x³-3x: punti stazionari in rosso")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** studia completamente $f(x)=\dfrac{x^2}{x-1}$.

**Ragionamento:**

**Passo 1 — dominio**: $x\ne1$ (notebook 3.3, 3.12).

**Passo 2 — simmetria**: $f(-x)=\dfrac{x^2}{-x-1}$, non uguale né a $f(x)$ né a $-f(x)$: nessuna simmetria.

**Passo 3 — asintoti.** Verticale: $x=1$ (il numeratore $1\ne0$ lì). Per l'asintoto obliquo (grado numeratore $=$ grado denominatore $+1$, notebook 3.12): dividiamo $x^2$ per $x-1$ (divisione polinomiale, notebook 3.12): $x^2=(x-1)(x+1)+1$, quindi $f(x)=x+1+\dfrac1{x-1}$. Per $x\to\pm\infty$, $\dfrac1{x-1}\to0$: asintoto obliquo $y=x+1$.

**Passo 4 — derivata prima.** Usando la forma $f(x)=x+1+\frac1{x-1}$ (più comoda da derivare, notebook 5.9):

$$f'(x) = 1-\frac1{(x-1)^2} = \frac{(x-1)^2-1}{(x-1)^2} = \frac{x^2-2x}{(x-1)^2} = \frac{x(x-2)}{(x-1)^2}$$

Studiamo il segno (notebook 1.9): il denominatore $(x-1)^2>0$ sempre (tranne in $x=1$, escluso dal dominio); il numeratore $x(x-2)$ è positivo per $x<0$ o $x>2$, negativo per $0<x<2$ (esclus $x=1$).

**Passo 5 — classifichiamo i punti stazionari** $x=0$ e $x=2$: $f'$ passa da $+$ a $-$ in $x=0$ (**massimo locale**, $f(0)=0$); da $-$ a $+$ in $x=2$ (**minimo locale**, $f(2)=4$).

**Passo 6 — sintesi**: $f$ cresce su $(-\infty,0)$, decresce su $(0,1)\cup(1,2)$, cresce su $(2,+\infty)$, con asintoto verticale $x=1$ e obliquo $y=x+1$.

**Perché riscrivere $f(x)$ nella forma "polinomio + resto" (dalla divisione) semplifica enormemente il calcolo della derivata:** derivare $\frac{x^2}{x-1}$ direttamente richiederebbe la regola del quoziente (notebook 5.9), con un'espressione più complicata da semplificare. Riscrivendo come $x+1+\frac1{x-1}$, la derivata si ottiene sommando le derivate dei pezzi (regola della somma, la più semplice), e il termine $\frac1{x-1}$ richiede solo la regola della catena applicata a $\frac1u$. È un esempio di come una manipolazione algebrica preliminare (qui, la divisione polinomiale) possa rendere molto più agevole il lavoro analitico successivo.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = x**2 / (x - 1)

f1 = sp.simplify(sp.diff(f, x))
print("f'(x) semplificata:", sp.factor(f1))

stazionari = sp.solve(sp.Eq(sp.diff(f,x), 0), x)
print("Punti stazionari:", stazionari)
for p in stazionari:
    print(f"  f({p}) = {f.subs(x, p)}")

# Asintoto obliquo
m = sp.limit(f/x, x, sp.oo)
q = sp.limit(f - m*x, x, sp.oo)
print(f"\nAsintoto obliquo: y = {m}x + {q}")
````
