## Teoria, passo dopo passo

### Convergenza assoluta

Una serie $\sum a_n$ **converge assolutamente** se $\sum|a_n|$ converge (una serie a termini non negativi, notebook 4.10). Un fatto fondamentale:

> Se $\sum a_n$ converge assolutamente, allora $\sum a_n$ converge.

**Dimostrazione** (con il criterio di Cauchy, notebook 5.3): sia $S_N=\sum_{n=0}^Na_n$ e $T_N=\sum_{n=0}^N|a_n|$. Se $(T_N)$ converge, è di Cauchy: per ogni $\varepsilon>0$ esiste $N$ tale che $|T_m-T_n|<\varepsilon$ per $m>n>N$, cioè $\sum_{k=n+1}^m|a_k|<\varepsilon$. Ma per la disuguaglianza triangolare (notebook 0.8, estesa a più termini):

$$|S_m-S_n| = \left|\sum_{k=n+1}^ma_k\right| \le \sum_{k=n+1}^m|a_k| < \varepsilon$$

Quindi anche $(S_N)$ è di Cauchy, dunque converge (criterio di Cauchy, notebook 5.3). ∎ La convergenza assoluta è quindi una condizione **più forte** e più facile da verificare (con i criteri del notebook 4.10, che richiedono termini non negativi) rispetto alla convergenza "semplice".

### Convergenza semplice ma non assoluta

Esistono serie che convergono, ma **non** assolutamente: si dicono a **convergenza condizionata**. L'esempio classico: $\sum\dfrac{(-1)^n}n$ (già incontrata nel notebook 4.11, al bordo del raggio di convergenza). La serie dei valori assoluti, $\sum\dfrac1n$, è la serie armonica, **divergente** (notebook 4.9): quindi $\sum\frac{(-1)^n}n$ non converge assolutamente. Eppure converge (semplicemente), per il criterio di Leibniz che segue.

### Il criterio di Leibniz per le serie a segni alterni

> Se $(b_n)$ è una successione **decrescente**, **positiva**, con $b_n\to0$, allora $\sum(-1)^nb_n$ **converge**.

**Intuizione della dimostrazione**: le somme parziali di indice pari $S_{2k}$ formano una successione decrescente e limitata inferiormente (da $S_1$); quelle di indice dispari $S_{2k+1}$ crescente e limitata superiormente (da $S_0$). Entrambe convergono (teorema delle successioni monotone, notebook 4.4-5.2), e la loro differenza $S_{2k}-S_{2k+1}=b_{2k+1}\to0$: quindi convergono allo **stesso** limite, e per un argomento sulle sottosuccessioni (notebook 4.6) l'intera successione delle somme parziali converge.

### Il criterio integrale (parallelo con gli integrali impropri)

Un ultimo collegamento, già anticipato nel notebook 5.18: se $f$ è positiva, decrescente, e $a_n=f(n)$, allora $\sum a_n$ converge se e solo se $\int_1^{+\infty}f(x)\,dx$ converge. Questo spiega perché $\sum\frac1{n^p}$ e $\int_1^{+\infty}\frac1{x^p}dx$ condividono **esattamente** la stessa soglia $p>1$ (notebook 4.10, 5.18): non è una coincidenza, ma una conseguenza diretta di questo criterio, che confronta l'area sotto una curva a scalini (la serie) con l'area sotto la curva continua corrispondente (l'integrale).

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

# Verifica: sum (-1)^n/n converge ma non assolutamente
def somma_parziale(termine, N):
    return sum(float(termine.subs(n, k)) for k in range(1, N+1))

termine_alterno = (-1)**n / n
for N in [10, 100, 1000, 10000]:
    S = somma_parziale(termine_alterno, N)
    print(f"N={N}: somma parziale = {S:.6f}")

print(f"\nValore atteso (convergenza nota): -ln(2) = {float(-sp.log(2)):.6f}")

# La serie dei valori assoluti (serie armonica) diverge
print("\nSerie dei valori assoluti (armonica): diverge, come noto dal notebook 4.9")

# Verifica del criterio integrale per p=2: confronto tra somma e integrale
x = sp.symbols('x', positive=True)
somma_p2 = sum(1/k**2 for k in range(1, 1000))
integrale_p2 = sp.integrate(1/x**2, (x, 1, sp.oo))
print(f"\nSomma parziale di sum 1/n^2 (N=1000): {float(somma_p2):.6f}")
print(f"Integrale di 1/x^2 da 1 a infinito: {integrale_p2}")
print(f"Somma esatta della serie (nota, pi^2/6): {float(sp.pi**2/6):.6f}")
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $\sum_{n=1}^\infty\dfrac{(-1)^n}{\sqrt n}$ converge, e se converge assolutamente.

**Ragionamento:**

**Passo 1 — verifichiamo la convergenza assoluta.** La serie dei valori assoluti è $\sum\dfrac1{\sqrt n}$, cioè $\sum\dfrac1{n^{1/2}}$: per il criterio del notebook 4.10 (e confermato dal criterio integrale di questo notebook), con $p=\frac12\le1$, questa serie **diverge**. Quindi $\sum\frac{(-1)^n}{\sqrt n}$ **non converge assolutamente**.

**Passo 2 — verifichiamo il criterio di Leibniz per la convergenza semplice.** Poniamo $b_n=\dfrac1{\sqrt n}$: dobbiamo verificare che sia decrescente, positiva, e tenda a $0$.

- **Positiva**: $b_n=\frac1{\sqrt n}>0$ per ogni $n\ge1$. ✓
- **Decrescente**: per $n<m$, $\sqrt n<\sqrt m$ (radice quadrata crescente, notebook 0.6), quindi $\frac1{\sqrt n}>\frac1{\sqrt m}$ (invertendo, notebook 1.9): $b_n$ è decrescente. ✓
- **Tende a $0$**: $\frac1{\sqrt n}\to0$ per $n\to\infty$ (notebook 4.2, con lo stesso schema ε-N già visto per $\frac1{n^2}$). ✓

**Passo 3 — applichiamo il criterio di Leibniz.** Tutte e tre le condizioni sono soddisfatte: per il criterio, $\sum(-1)^nb_n=\sum\dfrac{(-1)^n}{\sqrt n}$ **converge**.

**Conclusione**: la serie converge, ma **non assolutamente** — è un altro esempio di convergenza condizionata, come $\sum\frac{(-1)^n}n$ visto nella teoria.

**Perché è importante distinguere convergenza assoluta da condizionata, e non trattarle come equivalenti:** una serie a convergenza condizionata ha un comportamento "fragile": si può dimostrare (un risultato che non trattiamo per esteso, noto come **teorema di riarrangiamento di Riemann**) che, **riordinando** i suoi termini in modo diverso, si può far convergere la stessa serie a **qualunque** valore desiderato, o farla persino divergere! Questo è impossibile per una serie assolutamente convergente, il cui valore della somma è "stabile" rispetto a qualunque riordinamento. La distinzione, quindi, non è solo terminologica: riflette una differenza profonda nella robustezza del comportamento della serie.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

# Verifica della divergenza della serie dei valori assoluti (p=1/2)
somma_parziale_valori_assoluti = sum(1/k**0.5 for k in range(1, 10000))
print(f"Somma parziale di sum 1/sqrt(n) fino a N=10000: {somma_parziale_valori_assoluti:.4f}  (continua a crescere)")

# Verifica della convergenza della serie a segni alterni
somma_parziale_alterna = sum((-1)**k / k**0.5 for k in range(1, 10000))
print(f"Somma parziale di sum (-1)^n/sqrt(n) fino a N=10000: {somma_parziale_alterna:.6f}")

# Verifica delle tre condizioni del criterio di Leibniz
for k in [1, 2, 3, 10, 100]:
    b_k = 1 / k**0.5
    print(f"n={k}: b_n = {b_k:.4f}")
````

## Riepilogo finale

- **Convergenza assoluta** ($\sum|a_n|$ converge) implica convergenza semplice, dimostrabile con il criterio di Cauchy (notebook 5.3); il viceversa è falso.
- **Criterio di Leibniz**: se $(b_n)$ è decrescente, positiva, con $b_n\to0$, allora $\sum(-1)^nb_n$ converge — anche se non converge assolutamente (convergenza **condizionata**).
- Il **criterio integrale** collega serie e integrali impropri: $\sum f(n)$ e $\int_1^\infty f(x)dx$ condividono lo stesso comportamento (per $f$ positiva e decrescente), spiegando il parallelo tra $\sum\frac1{n^p}$ e $\int\frac1{x^p}dx$.
- Le serie a convergenza condizionata hanno un comportamento "instabile" rispetto al riordinamento dei termini, a differenza di quelle assolutamente convergenti.
