# 5.7 — Teoremi fondamentali sulle funzioni continue

**Prerequisiti:** 5.6 Continuità, 5.2 Completezza dei reali, 3.10 Funzioni limitate.


## Perché ci interessa

La continuità (notebook 5.6), unita alla completezza dei reali (notebook 5.2), ha conseguenze potentissime: su un intervallo chiuso e limitato, una funzione continua **deve** avere massimo e minimo, e **deve** assumere tutti i valori intermedi tra due suoi valori. Questi risultati — apparentemente "ovvi" guardando un grafico — richiedono in realtà la completezza per essere dimostrati, ed è per questo che falliscono su $\mathbb{Q}$ (dove certi valori "mancano").

## Teoria, passo dopo passo

### Teorema di Weierstrass (esistenza di massimo e minimo)

> Se $f$ è continua su un intervallo chiuso e limitato $[a,b]$, allora $f$ ammette **massimo** e **minimo** su $[a,b]$: esistono $x_M,x_m\in[a,b]$ tali che $f(x_m)\le f(x)\le f(x_M)$ per ogni $x\in[a,b]$.

**Perché servono entrambe le ipotesi (chiuso E limitato)**: su $(0,1)$ (aperto), $f(x)=x$ è continua ma non ha massimo (si avvicina a $1$ senza mai raggiungerlo, notebook 3.10, come $1-\frac1n$ nel notebook 5.1); su $[0,+\infty)$ (illimitato), $f(x)=x$ non ha massimo (cresce senza limite, notebook 3.11). Solo con **entrambe** le condizioni il teorema garantisce l'esistenza.

**Perché serve la continuità**: la funzione $f(x)=\frac1x$ per $x\in(0,1]$, con $f(0)$ definito artificialmente come $0$ (quindi $f:[0,1]\to\mathbb{R}$, dominio chiuso e limitato, ma **discontinua** in $0$, un asintoto, notebook 5.6), non ha massimo: cresce senza limite avvicinandosi a $0$.

**Intuizione della dimostrazione** (che non svolgiamo per intero, richiede strumenti più avanzati): la continuità su $[a,b]$ garantisce che $f$ sia limitata (altrimenti si potrebbe costruire, per assurdo, una successione senza sottosuccessioni convergenti in $[a,b]$, violando Bolzano-Weierstrass, notebook 4.6); essendo limitata, l'immagine ha un estremo superiore (completezza, notebook 5.1-5.2); la continuità garantisce che questo estremo superiore sia **effettivamente raggiunto** (non solo approssimato), diventando così un vero massimo.

### Teorema degli zeri (di Bolzano)

> Se $f$ è continua su $[a,b]$ e $f(a)\cdot f(b)<0$ (cioè $f(a)$ e $f(b)$ hanno segni opposti, notebook 0.2), allora esiste almeno un $c\in(a,b)$ con $f(c)=0$.

**Intuizione**: se il grafico parte sotto l'asse $x$ e finisce sopra (o viceversa), senza "saltare" (continuità), deve necessariamente attraversare l'asse da qualche parte. La dimostrazione rigorosa costruisce $c$ come estremo superiore dell'insieme $\{x\in[a,b] : f(x)<0\}$ (assumendo $f(a)<0<f(b)$): la completezza garantisce che questo estremo superiore esista, e la continuità garantisce che proprio lì $f$ valga $0$ (né positivo né negativo, per un argomento di permanenza del segno, notebook 5.3).

### Teorema dei valori intermedi

> Se $f$ è continua su $[a,b]$, allora $f$ assume **tutti** i valori compresi tra $f(a)$ e $f(b)$ (estremi inclusi).

Questo generalizza il teorema degli zeri: basta applicarlo alla funzione $g(x)=f(x)-k$ (per un valore intermedio $k$ qualsiasi tra $f(a)$ e $f(b)$), che è ancora continua (differenza di funzioni continue, notebook 5.6) e cambia segno tra $a$ e $b$ se $k$ è strettamente tra $f(a)$ e $f(b)$.

### Un'applicazione pratica: il metodo di bisezione

Il teorema degli zeri non è solo teorico: suggerisce un **algoritmo** per trovare (approssimativamente) uno zero. Se $f(a)<0<f(b)$, calcoliamo $f$ nel punto medio $m=\frac{a+b}2$ (notebook 2.3): se $f(m)=0$ abbiamo finito; altrimenti, $f(m)$ ha lo stesso segno di $f(a)$ o di $f(b)$, e possiamo "dimezzare" l'intervallo di ricerca, ripetendo. Questo produce una successione di intervalli sempre più piccoli, che si stringono attorno a uno zero — un'applicazione diretta e concreta del teorema.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Teorema degli zeri: verifica su f(x) = x^3 - x - 2
f = x**3 - x - 2
f_num = sp.lambdify(x, f, 'numpy')
print(f"f(1) = {f_num(1)}, f(2) = {f_num(2)}")
print(f"Cambiano segno? {f_num(1) * f_num(2) < 0}")

# Metodo di bisezione
def bisezione(funzione, a, b, tolleranza=1e-6, max_iter=50):
    for _ in range(max_iter):
        m = (a + b) / 2
        if abs(funzione(m)) < tolleranza or (b - a) / 2 < tolleranza:
            return m
        if funzione(a) * funzione(m) < 0:
            b = m
        else:
            a = m
    return (a + b) / 2

radice_approssimata = bisezione(f_num, 1, 2)
print(f"\nRadice approssimata (bisezione): {radice_approssimata:.6f}")

radice_esatta = sp.nsolve(f, x, 1.5)
print(f"Radice esatta (SymPy): {float(radice_esatta):.6f}")

# Teorema di Weierstrass: grafico che mostra massimo e minimo su [a,b]
x_vals = np.linspace(-2, 2, 200)
f2 = lambda t: t**3 - 3*t
fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(x_vals, f2(x_vals))
ax.axvline(x=-2, color='gray', linestyle='--')
ax.axvline(x=2, color='gray', linestyle='--')
massimo = max(f2(x_vals))
minimo = min(f2(x_vals))
ax.axhline(y=massimo, color='green', linestyle=':', label=f'massimo ≈ {massimo:.2f}')
ax.axhline(y=minimo, color='red', linestyle=':', label=f'minimo ≈ {minimo:.2f}')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("f(x)=x³-3x su [-2,2]: Weierstrass garantisce max e min")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dimostra che l'equazione $\cos x=x$ ha almeno una soluzione in $(0,1)$.

**Ragionamento:**

**Passo 1 — trasformiamo l'equazione in un problema di zeri.** Definiamo $g(x)=\cos x-x$: l'equazione $\cos x=x$ equivale a $g(x)=0$ (notebook 1.5, portando tutto da un lato).

**Passo 2 — verifichiamo la continuità.** $g$ è differenza di due funzioni continue ($\cos x$ e $x$, entrambe continue ovunque, notebook 5.6): $g$ è continua su tutto $\mathbb{R}$, in particolare su $[0,1]$.

**Passo 3 — calcoliamo $g$ agli estremi:**

$$g(0) = \cos(0)-0 = 1-0 = 1 > 0$$
$$g(1) = \cos(1)-1 \approx 0.540-1 = -0.460 < 0$$

**Passo 4 — applichiamo il teorema degli zeri.** $g$ è continua su $[0,1]$ e $g(0)\cdot g(1)<0$ (segni opposti): per il teorema, esiste $c\in(0,1)$ con $g(c)=0$, cioè $\cos c=c$.

**Conclusione**: l'equazione $\cos x=x$ ha almeno una soluzione in $(0,1)$ (questo valore, che si può approssimare con il metodo di bisezione, è noto come **costante di Dottie**, $\approx0.739$).

**Perché il teorema degli zeri è così utile per equazioni che non sappiamo risolvere algebricamente:** $\cos x=x$ mescola una funzione trigonometrica e una lineare — non esiste una formula chiusa per risolverla con le tecniche algebriche viste nei Moduli 1-2. Il teorema degli zeri non ci dà il valore esatto della soluzione, ma ne **garantisce l'esistenza** con assoluta certezza matematica, e il metodo di bisezione (teoria) permette poi di approssimarla numericamente quanto vogliamo — una strategia potentissima per equazioni "trascendenti" (che coinvolgono funzioni non polinomiali) prive di soluzione algebrica esplicita.

````python
import sympy as sp

x = sp.symbols('x', real=True)
g = sp.cos(x) - x

print(f"g(0) = {g.subs(x, 0)} = {float(g.subs(x, 0)):.4f}")
print(f"g(1) = {g.subs(x, 1)} ≈ {float(g.subs(x, 1)):.4f}")
print(f"Segni opposti? {g.subs(x,0) * g.subs(x,1) < 0}")

soluzione = sp.nsolve(sp.Eq(sp.cos(x), x), x, 0.5)
print(f"\nSoluzione approssimata (costante di Dottie): {float(soluzione):.6f}")
````

## Riepilogo finale

- **Teorema di Weierstrass**: una funzione continua su un intervallo **chiuso e limitato** ammette sempre massimo e minimo — entrambe le ipotesi (chiuso, limitato, continua) sono essenziali.
- **Teorema degli zeri**: se $f$ è continua su $[a,b]$ e cambia segno agli estremi, esiste uno zero in $(a,b)$.
- **Teorema dei valori intermedi**: generalizza il precedente a un valore intermedio $k$ qualsiasi tra $f(a)$ e $f(b)$.
- Il **metodo di bisezione** trasforma il teorema degli zeri in un algoritmo pratico per approssimare soluzioni di equazioni prive di formula chiusa.
