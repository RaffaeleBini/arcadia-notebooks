## Perché ci interessa

## Teoria, passo dopo passo

### La proprietà archimedea

> Per ogni $x\in\mathbb{R}$, esiste $n\in\mathbb{N}$ tale che $n>x$.

**Dimostrazione** (per assurdo, notebook 0.20): supponiamo che esista $x$ tale che $n\le x$ per **ogni** $n\in\mathbb{N}$. Allora $\mathbb{N}$ sarebbe limitato superiormente (da $x$), e per l'assioma di completezza (notebook 5.1) avrebbe un estremo superiore $L=\sup\mathbb{N}$. Ma allora $L-1$ non sarebbe un maggiorante di $\mathbb{N}$ (essendo $L$ il **minimo** dei maggioranti, $L-1<L$ non può esserlo): esiste $m\in\mathbb{N}$ con $m>L-1$, cioè $m+1>L$. Ma $m+1\in\mathbb{N}$, quindi $m+1\le L$ (essendo $L$ un maggiorante di $\mathbb{N}$): contraddizione ($m+1>L$ e $m+1\le L$ insieme). ∎

Questa dimostrazione usa **esattamente** lo schema del notebook 5.1: la completezza garantisce l'esistenza di $\sup\mathbb{N}$ (se $\mathbb{N}$ fosse limitato), e la contraddizione nasce dal fatto che nessun numero può essere il "minimo dei maggioranti" di un insieme che, aggiungendo $1$ a ciascun elemento, "genera sempre" elementi più grandi.

### Conseguenza: $\frac1n\to0$, finalmente rigoroso

Nel notebook 4.2 avevamo verificato $\frac1n\to0$ con la definizione ε-N, ma senza mai giustificare **perché** un tale $N$ esistesse sempre. Ora possiamo: dato $\varepsilon>0$, la proprietà archimedea garantisce l'esistenza di $N\in\mathbb{N}$ con $N>\frac1\varepsilon$, quindi $\frac1N<\varepsilon$ — esattamente la condizione richiesta. La proprietà archimedea è il "motore nascosto" che rende sempre possibile trovare l'indice $N$ richiesto dalla definizione di limite.

### Il teorema delle successioni monotone, dimostrato con rigore

Riprendiamo l'enunciato del notebook 4.4: se $(a_n)$ è crescente e limitata superiormente, converge. Ora possiamo dimostrarlo.

**Dimostrazione**: sia $A=\{a_n : n\in\mathbb{N}\}$ (l'immagine della successione, notebook 3.4), non vuoto e limitato superiormente per ipotesi. Per l'assioma di completezza, esiste $L=\sup A$. Dimostriamo che $a_n\to L$.

Fissiamo $\varepsilon>0$. Poiché $L=\sup A$ è il **minimo** dei maggioranti, $L-\varepsilon$ **non** è un maggiorante (essendo $L-\varepsilon<L$): esiste quindi un indice $N$ con $a_N>L-\varepsilon$. Ma $(a_n)$ è **crescente**: per ogni $n>N$, $a_n\ge a_N>L-\varepsilon$. Inoltre, essendo $L$ un maggiorante, $a_n\le L<L+\varepsilon$ per ogni $n$. Combinando:

$$L-\varepsilon < a_n < L+\varepsilon \qquad \text{per ogni } n>N$$

cioè $|a_n-L|<\varepsilon$ per ogni $n>N$: esattamente la definizione di limite (notebook 4.2). ∎

**Osservazione cruciale**: la dimostrazione mostra che il limite è **proprio** $L=\sup A$ — non un valore misterioso da scoprire, ma l'estremo superiore dell'immagine della successione, calcolabile in linea di principio con gli strumenti del notebook 5.1.

````python
import sympy as sp

# Verifica della proprietà archimedea: per ogni x, esiste n > x
def n_archimedeo(x):
    return sp.floor(x) + 1

for x_val in [5, 100, sp.pi, sp.E**10]:
    n = n_archimedeo(x_val)
    print(f"x={x_val}: n={n} > x? {n > x_val}")

# Applicazione del teorema dimostrato: successione a_n = 1 - 1/n (dal notebook 5.1)
n = sp.symbols('n', positive=True, integer=True)
a_n = 1 - 1/n

# L = sup A = 1 (trovato nel notebook 5.1). Verifichiamo la dimostrazione:
# per ogni epsilon, troviamo N tale che a_N > L - epsilon
def trova_N(epsilon, L=1):
    return sp.ceiling(1/epsilon)

for eps in [sp.Rational(1,10), sp.Rational(1,100)]:
    N = trova_N(eps)
    valore = a_n.subs(n, N)
    print(f"\neps={eps}: N={N}, a_N={valore}, L-eps={1-eps}, a_N > L-eps? {valore > 1-eps}")
````

## Esercizio svolto e commentato

**Problema:** usa il teorema delle successioni monotone (dimostrato in questo notebook) per dimostrare rigorosamente che la successione $a_1=\sqrt2$, $a_{n+1}=\sqrt{2+a_n}$ (già vista nel notebook 4.4) converge, identificando il limite come estremo superiore.

**Ragionamento:**

**Passo 1 — richiamiamo quanto già mostrato nel notebook 4.4** (per induzione, notebook 0.20): la successione è crescente e limitata superiormente da $2$. Quindi l'insieme $A=\{a_n:n\ge1\}$ è non vuoto e limitato superiormente.

**Passo 2 — applichiamo l'assioma di completezza** (notebook 5.1): esiste $L=\sup A\in\mathbb{R}$. Per il teorema dimostrato in questo notebook, $a_n\to L$.

**Passo 3 — troviamo $L$ esplicitamente**, passando al limite nella relazione ricorsiva (come già fatto, informalmente, nel notebook 4.4): $L=\sqrt{2+L}$, da cui $L^2-L-2=0$, cioè $(L-2)(L+1)=0$: $L=2$ (scartando $L=-1<0$, impossibile essendo $a_n>0$ sempre).

**Passo 4 — colleghiamo $L$ all'estremo superiore.** Ora sappiamo, con rigore, che $L=\sup A=2$: non è solo "il valore a cui sembra avvicinarsi la successione", ma precisamente il minimo dei maggioranti dell'insieme $\{a_n\}$ — un fatto che possiamo verificare indipendentemente con lo schema del notebook 5.1 (verificare che $2$ sia maggiorante, e che nessun numero più piccolo lo sia).

**Perché "passare al limite nella relazione ricorsiva" è ora un procedimento pienamente giustificato, e non solo un trucco algebrico:** nel notebook 4.4, questo passaggio (sostituire $L$ sia ad $a_n$ sia ad $a_{n+1}$ nella relazione ricorsiva) si basava sull'assunzione informale che il limite esistesse. Ora, avendo dimostrato **rigorosamente** l'esistenza del limite (teorema delle successioni monotone, con la completezza a garanzia), il passaggio al limite è un'operazione lecita in ogni suo passo: la successione $(a_{n+1})$ è una sottosuccessione (in realtà una "traslazione") di $(a_n)$, quindi condivide lo stesso limite $L$ (proprietà delle sottosuccessioni, notebook 4.6), permettendo di sostituire $L$ ovunque nella relazione ricorsiva senza alcuna ambiguità.

````python
import sympy as sp

L = sp.symbols('L', positive=True)
soluzioni = sp.solve(sp.Eq(L, sp.sqrt(2 + L)), L)
print("Soluzioni:", soluzioni)

# Verifica numerica di convergenza a L=2
def successione(N):
    a = sp.sqrt(2)
    valori = [a]
    for _ in range(N):
        a = sp.sqrt(2 + a)
        valori.append(a)
    return valori

termini = successione(6)
for i, t in enumerate(termini, start=1):
    print(f"a_{i} ≈ {float(t):.8f}")
````

## Riepilogo finale

- La **proprietà archimedea** (per ogni reale esiste un naturale più grande) è una conseguenza dimostrabile dell'assioma di completezza, non un fatto indipendente.
- Il **teorema delle successioni monotone** (notebook 4.4) è ora dimostrato con rigore: se $(a_n)$ è crescente e limitata, converge esattamente a $L=\sup\{a_n\}$.
- La dimostrazione usa lo schema "il limite meno epsilon non è un maggiorante" per trovare l'indice $N$ richiesto dalla definizione ε-N di limite.
- Questi strumenti (completezza, proprietà archimedea) sono il fondamento rigoroso su cui costruiremo, da qui in avanti, tutta l'analisi matematica del corso.
