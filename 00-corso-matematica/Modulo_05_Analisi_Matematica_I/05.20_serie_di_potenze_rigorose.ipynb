## Teoria, passo dopo passo

### Il raggio di convergenza, con la formula di Cauchy-Hadamard

Generalizziamo il calcolo del notebook 4.11 con una formula che funziona sempre (anche quando il criterio del rapporto non è applicabile direttamente):

$$\frac1R = \limsup_{n\to\infty}\sqrt[n]{|a_n|}$$

dove $\limsup$ (limite superiore) è un concetto che generalizza il limite per successioni che non convergono ma restano limitate (collegato alle sottosuccessioni, notebook 4.6): è il più grande valore a cui si "avvicinano infinite volte" i termini della successione. Non approfondiamo il $\limsup$ in dettaglio (un argomento per corsi più avanzati), ma nella maggior parte dei casi pratici questa formula coincide con il criterio del rapporto già usato nel notebook 4.11.

### Convergenza assoluta e uniforme dentro il raggio

All'interno del raggio di convergenza ($|x-x_0|<R$), una serie di potenze converge **assolutamente** (notebook 5.19): questo segue dal criterio del rapporto (o della radice) applicato a $\sum|a_n||x-x_0|^n$, con la stessa soglia $R$. Inoltre (un fatto più tecnico, che citiamo senza dimostrare) la convergenza è **uniforme** su ogni intervallo chiuso $[x_0-r,x_0+r]$ con $r<R$: questo garantisce che le operazioni di derivazione e integrazione si possano scambiare con la somma infinita, come vediamo ora.

### Derivazione e integrazione termine a termine

> Se $f(x)=\sum_{n=0}^\infty a_n(x-x_0)^n$ per $|x-x_0|<R$, allora $f$ è derivabile in quell'intervallo, e:
> $$f'(x) = \sum_{n=1}^\infty na_n(x-x_0)^{n-1}$$
> con lo **stesso** raggio di convergenza $R$. Analogamente, $f$ è integrabile termine a termine (teorema fondamentale, notebook 5.16), con lo stesso raggio.

**Perché questo risultato è potente**: possiamo trattare una serie di potenze "come se" fosse un polinomio (notebook 1.1), derivando o integrando ogni termine singolarmente — un'operazione che, per una somma infinita qualsiasi, non sarebbe affatto garantita senza la convergenza uniforme.

### Perché la serie di Taylor di $e^x$ converge davvero a $e^x$

Torniamo alla domanda iniziale. Per la formula di Taylor con resto di Lagrange (notebook 5.14):

$$e^x = P_n(x) + R_n(x), \qquad R_n(x)=\frac{e^\xi}{(n+1)!}x^{n+1}$$

Fissato $x$, e usando che $e^\xi\le e^{|x|}$ per $\xi$ tra $0$ e $x$ (una costante che **non dipende da $n$**): $|R_n(x)|\le\dfrac{e^{|x|}}{(n+1)!}|x|^{n+1}$. Ora, la successione $\dfrac{|x|^{n+1}}{(n+1)!}$ tende a $0$ per $n\to\infty$, **per ogni $x$ fissato** — un fatto dimostrabile applicando il criterio del rapporto (notebook 4.10) alla serie $\sum\frac{|x|^n}{n!}$, che risulta convergente (essa stessa una serie di potenze con $R=\infty$, notebook 4.11): se una serie converge, il suo termine generale tende a $0$ (condizione necessaria, notebook 4.7). Quindi $R_n(x)\to0$: il resto scompare, e $P_n(x)\to e^x$ — la serie di Taylor **converge davvero** alla funzione $e^x$, non solo per formalità algebrica. Questo tipo di verifica (mostrare che il resto tende a $0$) è ciò che rigorosamente giustifica ogni sviluppo in serie usato nel notebook 4.12.

````python
import sympy as sp

x, n = sp.symbols('x n', real=True)

# Verifica: la derivata termine a termine della serie di e^x ridà e^x
# (verifichiamo sul polinomio troncato, per illustrare)
serie_troncata = sum(x**k / sp.factorial(k) for k in range(10))
derivata_troncata = sp.diff(serie_troncata, x)
print("Serie troncata (10 termini) per e^x:", serie_troncata)
print("\nDerivata termine a termine:", sp.expand(derivata_troncata))
print("Confronto: coincide con la serie troncata a 9 termini (un ordine in meno)")

# Verifica che il resto di Taylor per e^x tenda a 0
def resto_stimato(x_val, ordine):
    M = float(sp.exp(abs(x_val)))
    return M * abs(x_val)**(ordine+1) / sp.factorial(ordine+1)

x_val = 2
for ordine in [5, 10, 15, 20]:
    R = resto_stimato(x_val, ordine)
    print(f"\nordine={ordine}: stima del resto R_n({x_val}) ≈ {float(R):.2e}")
````

## Esercizio svolto e commentato

**Problema:** usa la derivazione termine a termine per trovare la serie di potenze di $\dfrac1{(1-x)^2}$, a partire dalla serie geometrica $\sum x^n=\frac1{1-x}$ (notebook 4.8, 4.11).

**Ragionamento:**

**Passo 1 — riconosciamo che $\dfrac1{(1-x)^2}$ è la derivata di $\dfrac1{1-x}$** (verifichiamolo: $\frac{d}{dx}(1-x)^{-1}=-1\times(1-x)^{-2}\times(-1)=(1-x)^{-2}$, per la regola della catena, notebook 5.9).

**Passo 2 — applichiamo la derivazione termine a termine** (teoria) alla serie geometrica $\dfrac1{1-x}=\sum_{n=0}^\infty x^n$ (valida per $|x|<1$, notebook 4.8):

$$\frac1{(1-x)^2} = \frac{d}{dx}\sum_{n=0}^\infty x^n = \sum_{n=1}^\infty nx^{n-1}$$

(il termine $n=0$, essendo una costante, ha derivata $0$ e scompare dalla somma).

**Passo 3 — riscriviamo l'indice** per una forma più standard, ponendo $m=n-1$ (quindi $n=m+1$):

$$\frac1{(1-x)^2} = \sum_{m=0}^\infty(m+1)x^m = 1+2x+3x^2+4x^3+\dots$$

**Passo 4 — verifichiamo il raggio di convergenza**: per il teorema di derivazione termine a termine (teoria), resta $R=1$, lo stesso della serie di partenza.

**Verifica numerica** in $x=0.5$: $\dfrac1{(1-0.5)^2}=\dfrac1{0.25}=4$. Sommando i primi termini della serie: $1+2(0.5)+3(0.25)+4(0.125)+\dots=1+1+0.75+0.5+\dots$, che si avvicina a $4$ sommando più termini.

**Perché la derivazione termine a termine è così più efficiente del "ricominciare da capo" (per esempio, ricalcolare le derivate successive di $\frac1{(1-x)^2}$ una per una per la formula di Taylor):** avendo già a disposizione la serie (nota) di $\frac1{1-x}$, la derivazione termine a termine produce **immediatamente** la nuova serie con un'operazione puramente meccanica (derivare ogni $x^n$ separatamente, notebook 3.11), senza dover ricalcolare da zero infinite derivate successive della nuova funzione — un enorme risparmio di lavoro, reso possibile esattamente dal teorema di questo notebook.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica tramite serie troncata
serie_derivata = sum((n+1) * x**n for n in range(10))
print("Serie derivata troncata:", serie_derivata)

x_val = sp.Rational(1, 2)
somma_numerica = sum((n+1) * x_val**n for n in range(30))
valore_esatto = 1 / (1 - x_val)**2
print(f"\nSomma numerica (30 termini) in x=0.5: {float(somma_numerica):.6f}")
print(f"Valore esatto 1/(1-x)^2: {float(valore_esatto):.6f}")
````

<details>

**1.** $\sin x=\sum_{n=0}^\infty\frac{(-1)^nx^{2n+1}}{(2n+1)!}$. Derivando termine a termine: $\frac{d}{dx}\frac{(-1)^nx^{2n+1}}{(2n+1)!}=\frac{(-1)^n(2n+1)x^{2n}}{(2n+1)!}=\frac{(-1)^nx^{2n}}{(2n)!}$ — esattamente il termine generale della serie di $\cos x$ (notebook 4.12).

**2.** $\int-\sum x^n\,dx=-\sum\frac{x^{n+1}}{n+1}+c$. Imponendo che valga $\ln(1-0)=0$ in $x=0$: $c=0$. Quindi $\ln(1-x)=-\sum_{n=0}^\infty\frac{x^{n+1}}{n+1}=-\left(x+\frac{x^2}2+\frac{x^3}3+\dots\right)$.

**3.** Stesso raggio della serie di partenza (teoria: l'integrazione termine a termine preserva $R$): $R=1$.

**4.** $|R_n(x)|\le\dfrac{1}{(n+1)!}|x|^{n+1}$ (usando $M=1$, il maggiorante delle derivate di $\sin x$, notebook 5.14). Poiché $\sum\frac{|x|^n}{n!}$ converge per ogni $x$ (raggio infinito), il suo termine generale $\frac{|x|^{n+1}}{(n+1)!}\to0$ (condizione necessaria, notebook 4.7): quindi $R_n(x)\to0$ per ogni $x$ fissato, esattamente come per $e^x$ nella teoria.

## Riepilogo finale

- Il raggio di convergenza si calcola in generale con la formula di Cauchy-Hadamard, che nella pratica coincide quasi sempre con il criterio del rapporto (notebook 4.10-4.11).
- Dentro il raggio di convergenza, una serie di potenze converge assolutamente, e si può **derivare e integrare termine a termine**, mantenendo lo stesso raggio.
- La serie di Taylor di una funzione converge davvero alla funzione (non solo formalmente) quando il resto di Lagrange tende a $0$ — verificato esplicitamente per $e^x$ e (nell'esercizio) per $\sin x$.
- La derivazione/integrazione termine a termine permette di costruire nuove serie di potenze (come per $\frac1{(1-x)^2}$ o $\ln(1-x)$) a partire da serie già note, senza ricalcolare da zero.
