# 5.12 — Teorema di Cauchy

**Prerequisiti:** 5.11 Teorema di Lagrange, 3.6 Funzione inversa.


## Perché ci interessa

Il teorema di Lagrange (notebook 5.11) confronta la variazione di **una** funzione con la sua derivata. Il **teorema di Cauchy** (o degli incrementi finiti generalizzato) confronta le variazioni di **due** funzioni contemporaneamente — un risultato che sembra un dettaglio tecnico, ma è lo strumento chiave dietro un metodo che vedrai spesso nei corsi successivi per calcolare limiti di forme indeterminate (la regola di de l'Hôpital), e che qui usiamo per completare la catena Rolle → Lagrange → Cauchy.

## Teoria, passo dopo passo

### Enunciato

> Siano $f,g$ continue su $[a,b]$ e derivabili su $(a,b)$, con $g'(x)\ne0$ per ogni $x\in(a,b)$. Allora esiste $c\in(a,b)$ tale che:
> $$\frac{f'(c)}{g'(c)} = \frac{f(b)-f(a)}{g(b)-g(a)}$$

(la condizione $g'\ne0$ garantisce anche che $g(b)\ne g(a)$: se fosse $g(a)=g(b)$, il teorema di Rolle applicato a $g$ darebbe un punto con $g'=0$, contraddicendo l'ipotesi — quindi il denominatore a destra è sempre ben definito).

### Perché Lagrange è un caso particolare di Cauchy

Ponendo $g(x)=x$ (quindi $g'(x)=1\ne0$ sempre): il teorema di Cauchy diventa $\dfrac{f'(c)}1=\dfrac{f(b)-f(a)}{b-a}$, cioè $f'(c)=\dfrac{f(b)-f(a)}{b-a}$ — esattamente il teorema di Lagrange! La catena si completa: Rolle è un caso particolare di Lagrange (con $f(a)=f(b)$), e Lagrange è un caso particolare di Cauchy (con $g(x)=x$).

### Dimostrazione: di nuovo, ricondursi a Rolle

Definiamo la funzione ausiliaria:

$$h(x) = f(x) - f(a) - \frac{f(b)-f(a)}{g(b)-g(a)}\big(g(x)-g(a)\big)$$

(la stessa idea del notebook 5.11, ma "misurata" rispetto a $g$ invece che rispetto a $x$). Verifichiamo $h(a)=0$ (diretto, sostituendo $x=a$) e $h(b)=f(b)-f(a)-\frac{f(b)-f(a)}{g(b)-g(a)}(g(b)-g(a))=f(b)-f(a)-(f(b)-f(a))=0$. Per il teorema di Rolle applicato a $h$ (continua su $[a,b]$, derivabile su $(a,b)$, con $h(a)=h(b)=0$): esiste $c\in(a,b)$ con $h'(c)=0$, cioè:

$$f'(c) - \frac{f(b)-f(a)}{g(b)-g(a)}g'(c) = 0 \quad\Rightarrow\quad f'(c) = \frac{f(b)-f(a)}{g(b)-g(a)}g'(c)$$

Dividendo per $g'(c)$ (lecito, $g'(c)\ne0$ per ipotesi): $\dfrac{f'(c)}{g'(c)}=\dfrac{f(b)-f(a)}{g(b)-g(a)}$. ∎

### Un'applicazione: verso la regola di de l'Hôpital

Il teorema di Cauchy è il cuore della dimostrazione della **regola di de l'Hôpital**, uno strumento (che approfondirai in corsi successivi) per calcolare limiti di forme indeterminate $\frac00$ o $\frac\infty\infty$ (notebook 4.3) confrontando le derivate di numeratore e denominatore invece delle funzioni stesse. L'idea, in breve: se $f(a)=g(a)=0$, applicando Cauchy su $[a,x]$ (con $x\to a$) si ottiene $\dfrac{f(x)}{g(x)}=\dfrac{f(x)-f(a)}{g(x)-g(a)}=\dfrac{f'(c)}{g'(c)}$ per qualche $c$ tra $a$ e $x$: quando $x\to a$, anche $c\to a$ (essendo "schiacciato" tra $a$ e $x$), suggerendo $\lim_{x\to a}\dfrac{f(x)}{g(x)}=\lim_{x\to a}\dfrac{f'(x)}{g'(x)}$, se quest'ultimo limite esiste.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica del teorema di Cauchy con f(x)=x^2, g(x)=x^3 su [1,2]
f = x**2
g = x**3
a_val, b_val = 1, 2

rapporto_destro = (f.subs(x,b_val) - f.subs(x,a_val)) / (g.subs(x,b_val) - g.subs(x,a_val))
print(f"Rapporto (f(b)-f(a))/(g(b)-g(a)) = {rapporto_destro}")

f_prime, g_prime = sp.diff(f, x), sp.diff(g, x)
rapporto_c = f_prime / g_prime
c_soluzioni = sp.solve(sp.Eq(rapporto_c, rapporto_destro), x)
print(f"Soluzioni per c: {c_soluzioni}")
c_valido = [c for c in c_soluzioni if a_val < c < b_val]
print(f"Punto valido in (1,2): {c_valido}")

# Verifica che Lagrange sia il caso g(x)=x del teorema di Cauchy
g_identita = x
print(f"\nCon g(x)=x: g'(x) = {sp.diff(g_identita, x)}  (sempre 1, coerente con Lagrange)")
````

## Esercizio svolto e commentato

**Problema:** applica il teorema di Cauchy a $f(x)=\sin x$, $g(x)=\cos x$ su $\left[0,\dfrac\pi2\right]$, e trova esplicitamente il punto $c$.

**Ragionamento:**

**Passo 1 — verifichiamo le ipotesi.** $f,g$ sono continue e derivabili ovunque (notebook 2.9, 5.9). Calcoliamo $g'(x)=-\sin x$: si annulla solo in $x=0,\pi,\dots$ — nell'intervallo **aperto** $\left(0,\frac\pi2\right)$, $g'(x)\ne0$ (essendo $\sin x>0$ per $x$ strettamente tra $0$ e $\pi$). ✓

**Passo 2 — calcoliamo il rapporto a destra:**

$$\frac{f(\pi/2)-f(0)}{g(\pi/2)-g(0)} = \frac{\sin(\pi/2)-\sin(0)}{\cos(\pi/2)-\cos(0)} = \frac{1-0}{0-1} = -1$$

**Passo 3 — impostiamo l'equazione per $c$:**

$$\frac{f'(c)}{g'(c)} = \frac{\cos c}{-\sin c} = -\cot c$$

Vogliamo $-\cot c=-1$, cioè $\cot c=1$, cioè $\dfrac{\cos c}{\sin c}=1$, cioè $\tan c=1$ (essendo $\cot c$ il reciproco di $\tan c$).

**Passo 4 — risolviamo** (valore notevole, notebook 2.9): $\tan c=1 \Rightarrow c=\dfrac\pi4$, che appartiene a $\left(0,\dfrac\pi2\right)$. ✓

**Verifica diretta**: $f'\left(\frac\pi4\right)=\cos\frac\pi4=\frac{\sqrt2}2$, $g'\left(\frac\pi4\right)=-\sin\frac\pi4=-\frac{\sqrt2}2$. Rapporto: $\dfrac{\sqrt2/2}{-\sqrt2/2}=-1$. ✓ Coincide con il rapporto calcolato al passo 2.

**Perché il punto $c=\frac\pi4$ è particolarmente "elegante" in questo esempio, e cosa suggerisce sulla scelta delle funzioni:** avendo scelto $f=\sin$ e $g=\cos$, legate dalla stessa identità fondamentale (notebook 2.9) e dalla stessa periodicità, il punto medio dell'intervallo $\left[0,\frac\pi2\right]$ risulta essere anche il punto di simmetria per entrambe le funzioni (dove $\sin=\cos=\frac{\sqrt2}2$): non stupisce che proprio lì si verifichi la condizione del teorema. Con funzioni meno simmetriche, il punto $c$ non avrebbe necessariamente una forma così riconoscibile.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f, g = sp.sin(x), sp.cos(x)

rapporto_destro = (f.subs(x, sp.pi/2) - f.subs(x, 0)) / (g.subs(x, sp.pi/2) - g.subs(x, 0))
print("Rapporto destro:", rapporto_destro)

f_prime, g_prime = sp.diff(f, x), sp.diff(g, x)
c_soluzioni = sp.solveset(sp.Eq(f_prime/g_prime, rapporto_destro), x, domain=sp.Interval.open(0, sp.pi/2))
print("Soluzione per c:", c_soluzioni)
````

## Riepilogo finale

- **Teorema di Cauchy**: $f,g$ continue su $[a,b]$, derivabili su $(a,b)$, con $g'\ne0$ su $(a,b)$, garantisce $c$ con $\dfrac{f'(c)}{g'(c)}=\dfrac{f(b)-f(a)}{g(b)-g(a)}$.
- Ponendo $g(x)=x$, si ritrova esattamente il teorema di Lagrange: la catena Rolle → Lagrange → Cauchy è una successione di generalizzazioni, ciascuna dimostrata riconducendosi alla precedente.
- La dimostrazione usa la stessa strategia delle due precedenti: costruire una funzione ausiliaria che si annulla agli estremi, poi applicare Rolle.
- Il teorema di Cauchy è lo strumento alla base della regola di de l'Hôpital per il calcolo di limiti in forma indeterminata.
