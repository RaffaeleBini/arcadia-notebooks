# 5.5 — Limiti notevoli

**Prerequisiti:** 5.4 Limiti, 2.9 Seno, coseno, tangente, 4.4 Successioni monotone.


## Perché ci interessa

Alcuni limiti si ripresentano così spesso, e sono così importanti come "mattoni" per calcolarne altri, che vale la pena conoscerli a memoria — proprio come i valori notevoli di seno e coseno (notebook 2.9). Questi **limiti notevoli** collegano trigonometria, esponenziali e logaritmi in modo sorprendente, e prepareranno il terreno per la derivata (notebook 5.8), che si definisce proprio come un limite.

## Teoria, passo dopo passo

### Il limite fondamentale: $\lim_{x\to0}\dfrac{\sin x}x=1$

Questo è probabilmente il limite notevole più importante. **Dimostrazione geometrica** (per $0<x<\frac\pi2$, usando la circonferenza goniometrica, notebook 2.8-2.9): confrontiamo le aree di tre figure costruite sull'angolo $x$ — il triangolo inscritto (area $\frac12\sin x$), il settore circolare (area $\frac12x$), il triangolo circoscritto (area $\frac12\tan x$). Geometricamente, queste aree sono in ordine crescente:

$$\frac12\sin x \le \frac12x \le \frac12\tan x$$

Moltiplicando per $\frac2{\sin x}$ (positivo per $0<x<\frac\pi2$, il verso non si inverte, notebook 1.8):

$$1 \le \frac{x}{\sin x} \le \frac1{\cos x}$$

Invertendo tutto (per quantità positive, l'ordine si inverte, notebook 1.9):

$$\cos x \le \frac{\sin x}x \le 1$$

Quando $x\to0$, $\cos x\to1$ (continuità del coseno, che vedremo nel notebook 5.6): la funzione $\frac{\sin x}x$ è "schiacciata" tra due funzioni che tendono entrambe a $1$. Questo argomento — noto come **teorema del confronto** (o "dei due carabinieri") — garantisce $\lim_{x\to0}\frac{\sin x}x=1$.

### Limiti derivati dal precedente

Da $\frac{\sin x}x\to1$ si ottengono, con manipolazioni algebriche, altri limiti notevoli:

$$\lim_{x\to0}\frac{1-\cos x}{x^2} = \frac12$$

(si dimostra moltiplicando numeratore e denominatore per $1+\cos x$, una razionalizzazione analoga a quella vista nel notebook 5.4, esercizio 4, e usando l'identità $1-\cos^2x=\sin^2x$, notebook 2.9).

$$\lim_{x\to0}\frac{\tan x}x = 1$$

(scrivendo $\tan x=\frac{\sin x}{\cos x}$ e usando l'algebra dei limiti, notebook 4.3, ora rigorosa anche per funzioni).

### Il limite che definisce $e$, ora per funzioni

Ricordiamo dal notebook 4.4 che $e=\lim_{n\to\infty}\left(1+\frac1n\right)^n$ (limite di successione). La versione per funzioni (con $x$ reale, non solo naturale):

$$\lim_{x\to+\infty}\left(1+\frac1x\right)^x = e \qquad\qquad \lim_{x\to0}(1+x)^{1/x} = e$$

(le due forme sono equivalenti, sostituendo $x=\frac1t$). Da questo derivano limiti collegati a esponenziali e logaritmi:

$$\lim_{x\to0}\frac{\ln(1+x)}x = 1 \qquad\qquad \lim_{x\to0}\frac{e^x-1}x = 1$$

Questi ultimi due limiti anticipano, in modo diretto, il valore della **derivata** di $\ln x$ e $e^x$ nel punto $0$ (o $1$): il Modulo 5 li ritroverà esattamente in questa forma quando definiremo la derivata come limite del rapporto incrementale (notebook 5.8).

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica dei limiti notevoli con SymPy
limiti_notevoli = {
    "sin(x)/x": sp.sin(x)/x,
    "(1-cos(x))/x^2": (1 - sp.cos(x)) / x**2,
    "tan(x)/x": sp.tan(x) / x,
    "ln(1+x)/x": sp.log(1+x) / x,
    "(e^x - 1)/x": (sp.exp(x) - 1) / x,
}

for nome, espressione in limiti_notevoli.items():
    limite = sp.limit(espressione, x, 0)
    print(f"lim_(x->0) {nome} = {limite}")

# Verifica numerica del teorema del confronto per sin(x)/x
import numpy as np
x_vals = np.linspace(0.01, 1.5, 5)
for xv in x_vals:
    cos_val = np.cos(xv)
    rapporto = np.sin(xv) / xv
    print(f"\nx={xv:.2f}: cos(x)={cos_val:.4f} <= sin(x)/x={rapporto:.4f} <= 1")
````

## Esercizio svolto e commentato

**Problema:** calcola $\lim_{x\to0}\dfrac{\sin(3x)}{x}$, usando il limite notevole fondamentale (senza usare direttamente SymPy per il risultato finale).

**Ragionamento:**

**Passo 1 — riconosciamo che l'argomento del seno non è $x$, ma $3x$**: non possiamo applicare direttamente $\frac{\sin x}x\to1$, serve un piccolo aggiustamento algebrico.

**Passo 2 — moltiplichiamo e dividiamo per $3$**, per far comparire la forma $\frac{\sin(3x)}{3x}$:

$$\frac{\sin(3x)}x = 3\cdot\frac{\sin(3x)}{3x}$$

**Passo 3 — operiamo un cambio di variabile mentale**: poniamo $t=3x$. Quando $x\to0$, anche $t\to0$ (essendo $t$ un multiplo costante di $x$). Quindi:

$$\frac{\sin(3x)}{3x} = \frac{\sin t}t \ \xrightarrow[t\to0]{} \ 1$$

(esattamente il limite notevole fondamentale, applicato alla variabile $t$ invece che $x$ — il nome della variabile non conta, solo il fatto che tenda a $0$).

**Passo 4 — riportiamo il fattore $3$ messo da parte al passo 2:**

$$\lim_{x\to0}\frac{\sin(3x)}x = 3\times1 = 3$$

**Perché questa tecnica di "moltiplicare e dividere per il coefficiente dell'argomento" funziona sempre per limiti del tipo $\frac{\sin(kx)}x$:** il limite notevole fondamentale richiede che l'argomento del seno e il denominatore siano **esattamente gli stessi**. Quando non lo sono (qui, $3x$ contro $x$), creiamo artificialmente quella corrispondenza moltiplicando e dividendo per il fattore mancante ($3$), aggiustando poi il risultato con lo stesso fattore fuori dal limite. In generale, $\lim_{x\to0}\frac{\sin(kx)}x=k$, per qualunque costante $k\ne0$.

````python
import sympy as sp

x = sp.symbols('x', real=True)
espressione = sp.sin(3*x) / x
limite = sp.limit(espressione, x, 0)
print("Limite:", limite)

# Verifica della tecnica passo-passo
t = sp.symbols('t', real=True)
verifica = 3 * sp.limit(sp.sin(t)/t, t, 0)
print("Verifica con la tecnica (3 * lim sin(t)/t):", verifica)
````

## Riepilogo finale

- $\lim_{x\to0}\frac{\sin x}x=1$: dimostrato con il teorema del confronto, applicato a un'ineguaglianza tra aree costruite sulla circonferenza goniometrica.
- Da questo derivano $\frac{1-\cos x}{x^2}\to\frac12$ e $\frac{\tan x}x\to1$.
- $\left(1+\frac1x\right)^x\to e$ (per $x\to\infty$), da cui $\frac{\ln(1+x)}x\to1$ e $\frac{e^x-1}x\to1$ (per $x\to0$).
- La tecnica di "moltiplicare e dividere" per adattare un'espressione alla forma di un limite notevole noto è lo strumento chiave per calcolare limiti più complessi a partire da questi mattoni fondamentali.
