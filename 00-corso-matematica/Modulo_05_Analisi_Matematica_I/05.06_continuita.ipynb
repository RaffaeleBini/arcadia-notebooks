# 5.6 — Continuità

**Prerequisiti:** 5.4 Limiti, 3.12 Funzioni razionali.


## Perché ci interessa

Intuitivamente, una funzione "continua" è una che si può disegnare senza staccare la penna dal foglio: nessun salto improvviso. Nel notebook 3.12 abbiamo già incontrato funzioni con "buchi" (discontinuità eliminabili) o asintoti verticali (dove la funzione "esplode"). Ora, avendo a disposizione la definizione rigorosa di limite (notebook 5.4), possiamo formalizzare con precisione cosa significa continuità — un concetto che sarà l'ipotesi di quasi tutti i teoremi importanti del resto del modulo.

## Teoria, passo dopo passo

### Definizione

$f$ è **continua in $x_0$** se sono soddisfatte **tre condizioni**:

1. $f(x_0)$ è definito (cioè $x_0$ appartiene al dominio di $f$, notebook 3.3);
2. $\lim_{x\to x_0}f(x)$ esiste (finito);
3. $\lim_{x\to x_0}f(x) = f(x_0)$ (il limite coincide con il valore effettivo della funzione).

In modo compatto, usando la definizione ε-δ (notebook 5.4): per ogni $\varepsilon>0$ esiste $\delta>0$ tale che $|x-x_0|<\delta \Rightarrow |f(x)-f(x_0)|<\varepsilon$ (qui **non** serve escludere $x=x_0$ con la condizione $0<|x-x_0|$, perché quando $x=x_0$ la disuguaglianza $|f(x_0)-f(x_0)|=0<\varepsilon$ è automaticamente vera).

$f$ è **continua su un intervallo** se è continua in ogni suo punto.

### Le funzioni elementari sono continue nel loro dominio

Un fatto che diamo per buono (dimostrabile con la definizione ε-δ, uno per uno): polinomi (notebook 3.11), funzioni razionali (notebook 3.12, nei punti dove sono definite), radici, esponenziali, logaritmi, funzioni trigonometriche (Modulo 2-3) sono tutte continue in ogni punto del proprio dominio naturale. Questo giustifica perché, per queste funzioni, si può spesso calcolare un limite semplicemente **sostituendo** il valore (come abbiamo fatto informalmente fin dal Modulo 4): $\lim_{x\to x_0}f(x)=f(x_0)$, esattamente la definizione di continuità.

### Tipi di discontinuità

Riprendiamo e classifichiamo con precisione i comportamenti "irregolari" già intravisti nel notebook 3.12:

- **Discontinuità eliminabile**: il limite $\lim_{x\to x_0}f(x)$ esiste (finito), ma o $f(x_0)$ non è definito, o è definito con un valore diverso dal limite. Esempio: $f(x)=\frac{x^2-1}{x-1}$ in $x_0=1$ (notebook 3.12): il limite vale $2$, ma $f(1)$ non è definito. Si chiama "eliminabile" perché basterebbe **ridefinire** $f(1)=2$ per rendere la funzione continua lì.
- **Discontinuità di salto**: i limiti destro e sinistro (notebook 5.4) esistono entrambi finiti, ma sono **diversi**. Esempio: $f(x)=\frac{|x|}x$ in $x_0=0$ (notebook 5.4): limite destro $1$, sinistro $-1$.
- **Discontinuità di seconda specie** (o essenziale): almeno uno dei due limiti unilaterali non esiste, oppure è infinito. Esempio: $f(x)=\frac1x$ in $x_0=0$ (limiti $\pm\infty$, un asintoto verticale, notebook 3.12).

### Operazioni che preservano la continuità

Se $f$ e $g$ sono continue in $x_0$, allora $f+g$, $f-g$, $f\cdot g$ sono continue in $x_0$; $\frac fg$ è continua in $x_0$ se inoltre $g(x_0)\ne0$. Anche la **composizione** (notebook 3.7) di funzioni continue è continua: se $g$ è continua in $x_0$ e $f$ è continua in $g(x_0)$, allora $f\circ g$ è continua in $x_0$. Queste proprietà seguono direttamente dall'algebra dei limiti (notebook 4.3, 5.3), ora applicata a funzioni invece che a successioni.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica dei tre tipi di discontinuità in x_0 = 0 (o 1)

# 1. Discontinuità eliminabile: f(x) = (x^2-1)/(x-1) in x=1
f1 = (x**2 - 1) / (x - 1)
limite1 = sp.limit(f1, x, 1)
print(f"Discontinuità eliminabile: lim f(x) = {limite1}, ma f(1) non è definito (0/0)")

# 2. Discontinuità di salto: f(x) = |x|/x in x=0
f2 = sp.Abs(x) / x
lim_dx = sp.limit(f2, x, 0, '+')
lim_sx = sp.limit(f2, x, 0, '-')
print(f"\nDiscontinuità di salto: lim destro = {lim_dx}, lim sinistro = {lim_sx}")

# 3. Discontinuità di seconda specie: f(x) = 1/x in x=0
f3 = 1/x
lim_dx3 = sp.limit(f3, x, 0, '+')
lim_sx3 = sp.limit(f3, x, 0, '-')
print(f"\nDiscontinuità essenziale: lim destro = {lim_dx3}, lim sinistro = {lim_sx3}")

# Verifica che i polinomi siano continui (il limite coincide sempre con la sostituzione diretta)
f4 = x**3 - 2*x + 1
x0 = 2
print(f"\nPolinomio continuo: lim_(x->{x0}) f(x) = {sp.limit(f4, x, x0)}, f({x0}) = {f4.subs(x, x0)}")
````

## Esercizio svolto e commentato

**Problema:** studia la continuità in $x_0=0$ della funzione a tratti:

$$f(x) = \begin{cases} \dfrac{\sin x}x & \text{se } x\ne0 \\ 1 & \text{se } x=0 \end{cases}$$

**Ragionamento:**

**Passo 1 — verifichiamo la prima condizione della definizione**: $f(0)$ è definito? Sì, per la regola a tratti (notebook 1.14), $f(0)=1$.

**Passo 2 — verifichiamo la seconda condizione**: esiste $\lim_{x\to0}f(x)$? Per $x\ne0$, $f(x)=\frac{\sin x}x$, e conosciamo questo limite dal notebook 5.5 (il limite notevole fondamentale): $\lim_{x\to0}\frac{\sin x}x=1$. Il limite esiste, e vale $1$.

**Passo 3 — verifichiamo la terza condizione**: il limite coincide con $f(0)$? Sì: $\lim_{x\to0}f(x)=1=f(0)$.

**Conclusione**: tutte e tre le condizioni sono soddisfatte: $f$ è **continua** in $x_0=0$.

**Perché questa funzione è stata definita "apposta" con $f(0)=1$, e non con un altro valore:** l'espressione $\frac{\sin x}x$ da sola **non è definita** in $x=0$ (divisione per zero, notebook 0.3): senza la regola a tratti, avremmo una discontinuità eliminabile (esattamente come nell'esempio $\frac{x^2-1}{x-1}$ della teoria). Scegliendo di **definire** $f(0)$ esattamente uguale al valore del limite (che sappiamo essere $1$, dal limite notevole), "ripariamo" artificialmente il buco, ottenendo una funzione continua ovunque — questa è precisamente l'operazione descritta come "eliminare" una discontinuità eliminabile: ridefinire il valore nel punto mancante in modo che coincida con il limite.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Definiamo f a tratti con sympy
f = sp.Piecewise((sp.sin(x)/x, x != 0), (1, True))

limite = sp.limit(sp.sin(x)/x, x, 0)
print("Limite per x->0 di sin(x)/x:", limite)
print("Valore di f in 0 (definito a tratti):", 1)
print("Coincidono? La funzione è continua in 0:", limite == 1)
````

## Riepilogo finale

- $f$ è continua in $x_0$ se $f(x_0)$ è definito, il limite $\lim_{x\to x_0}f(x)$ esiste finito, e i due coincidono.
- Tre tipi di discontinuità: **eliminabile** (limite finito ma diverso da $f(x_0)$, o $f(x_0)$ non definito), **di salto** (limiti unilaterali finiti ma diversi), **di seconda specie** (almeno un limite unilaterale infinito o non esistente).
- Le funzioni elementari (polinomi, razionali, esponenziali, logaritmi, trigonometriche) sono continue nel loro dominio naturale: per esse, calcolare il limite equivale a sostituire il valore.
- Somma, prodotto, quoziente (dove definito) e composizione di funzioni continue sono continue.
