Questo notebook apre il Modulo 5 con l'approccio opposto: invece di costruire $\mathbb{R}$ dal basso, lo **caratterizziamo dall'alto** con un elenco di proprietà (assiomi) che lo definiscono univocamente. È il punto di partenza di tutta l'analisi rigorosa che seguirà.

## Teoria, passo dopo passo

### Cosa significa "assiomatizzare"

Un **assioma** è un'affermazione che si assume vera senza dimostrazione, come punto di partenza per costruire una teoria (un'idea che abbiamo incontrato implicitamente fin dal notebook 0.20, parlando di dimostrazioni: bisogna sempre partire da qualcosa). Invece di dire "$\mathbb{R}$ è ciò che si ottiene aggiungendo gli irrazionali a $\mathbb{Q}$", diciamo: "$\mathbb{R}$ è **l'unico** insieme che soddisfa queste proprietà" — e da queste proprietà deriviamo tutto il resto per dimostrazione, con lo stesso rigore visto nel notebook 0.20.

### Assiomi di campo (proprietà algebriche)

$\mathbb{R}$, con le operazioni $+$ e $\times$, soddisfa gli **assiomi di campo**: essenzialmente, tutte le proprietà algebriche che usiamo dal notebook 0.5 senza più pensarci — commutativa, associativa, distributiva, esistenza di elemento neutro e opposto per la somma, esistenza di elemento neutro e reciproco per il prodotto (tranne che per $0$). Queste proprietà, prese da sole, sono soddisfatte anche da $\mathbb{Q}$: non bastano a distinguere $\mathbb{R}$ da $\mathbb{Q}$.

### Assiomi di ordine

$\mathbb{R}$ è **totalmente ordinato**: per ogni $x,y\in\mathbb{R}$ vale esattamente una tra $x<y$, $x=y$, $x>y$ (tricotomia, già usata implicitamente nel notebook 3.9), e l'ordine è compatibile con le operazioni (se $x<y$ allora $x+z<y+z$; se $x<y$ e $z>0$ allora $xz<yz$ — le stesse regole delle disequazioni del notebook 1.8). Anche queste proprietà valgono per $\mathbb{Q}$: ancora non basta a distinguerlo da $\mathbb{R}$.

### L'assioma di completezza: ciò che distingue $\mathbb{R}$ da $\mathbb{Q}$

L'ultimo assioma è quello decisivo. Prima serve una definizione: un insieme $A\subseteq\mathbb{R}$ è **limitato superiormente** (notebook 3.10) se esiste $M$ con $a\le M$ per ogni $a\in A$. Un **estremo superiore** di $A$ è il **minimo** tra tutti i maggioranti — se esiste.

> **Assioma di completezza**: ogni sottoinsieme di $\mathbb{R}$ non vuoto e limitato superiormente ammette estremo superiore (in $\mathbb{R}$).

Questo è **falso** in $\mathbb{Q}$! Consideriamo $A=\{x\in\mathbb{Q} : x^2<2\}$: è limitato superiormente (per esempio da $2$), ma **non ha estremo superiore in $\mathbb{Q}$** — il candidato naturale sarebbe $\sqrt2$, che (notebook 0.4) non è razionale. In $\mathbb{Q}$, l'insieme $A$ ha un "buco" esattamente dove dovrebbe stare il suo estremo superiore. In $\mathbb{R}$, invece, $\sup A=\sqrt2$ esiste sempre, per l'assioma.

### Perché questo assioma "riempie" davvero tutti i buchi

L'assioma di completezza è precisamente ciò che rende rigoroso il teorema delle successioni monotone (notebook 4.4): una successione crescente e limitata superiormente ha come limite esattamente l'estremo superiore del suo insieme di valori (l'immagine, notebook 3.4) — un fatto che potremo finalmente dimostrare con precisione nel notebook 5.2, invece di limitarci all'intuizione geometrica.

````python
import sympy as sp

# Verifichiamo che l'insieme A = {x in Q : x^2 < 2} non ha massimo razionale
# (approssimando sqrt(2) sempre meglio con razionali)
import fractions

def approssima_sqrt2(precisione):
    x = fractions.Fraction(1)
    for _ in range(50):
        x = (x + 2/x) / 2  # metodo di Newton, converge a sqrt(2)
    return x

approssimazione = approssima_sqrt2(50)
print(f"Approssimazione razionale di sqrt(2): {float(approssimazione):.15f}")
print(f"sqrt(2) reale: {float(sp.sqrt(2)):.15f}")
print(f"(approssimazione)^2 = {float(approssimazione**2):.15f}  (mai esattamente 2, essendo razionale)")

# Per ogni razionale q con q^2<2, esiste un razionale piu' grande q' con (q')^2<2 ancora
# Questo mostra che A non ha massimo: l'estremo superiore "dovrebbe" essere sqrt(2), ma non è in Q
q = sp.Rational(14, 10)
print(f"\nq={q}, q^2={q**2} < 2? {q**2 < 2}")
q_piu_grande = sp.Rational(141, 100)
print(f"q'={q_piu_grande}, (q')^2={q_piu_grande**2} < 2? {q_piu_grande**2 < 2}  (q' > q, ancora nell'insieme)")
````

## Esercizio svolto e commentato

**Problema:** trova l'estremo superiore di $A=\left\{1-\dfrac1n : n\in\mathbb{N},\ n\ge1\right\}$, e verifica che soddisfi la definizione (minimo tra i maggioranti).

**Ragionamento:**

**Passo 1 — elenchiamo alcuni elementi**: $n=1\Rightarrow0$; $n=2\Rightarrow\frac12$; $n=3\Rightarrow\frac23$; $n=10\Rightarrow\frac9{10}$; $n=100\Rightarrow\frac{99}{100}$. Gli elementi si avvicinano sempre di più a $1$, restando sempre **strettamente minori** di $1$.

**Passo 2 — congetturiamo $\sup A=1$, e verifichiamo che $1$ sia un maggiorante.** Per ogni $n\ge1$: $\frac1n>0$, quindi $1-\frac1n<1$: sì, $1$ è un maggiorante di $A$.

**Passo 3 — verifichiamo che sia il **minimo** dei maggioranti** (la parte più delicata della definizione di estremo superiore). Dobbiamo mostrare che **nessun** numero minore di $1$ può essere un maggiorante: preso un qualunque $M<1$, dobbiamo trovare un elemento di $A$ che superi $M$.

Sia $M<1$. Cerchiamo $n$ tale che $1-\frac1n>M$, cioè $\frac1n<1-M$, cioè $n>\frac1{1-M}$ (qui usiamo che $1-M>0$, essendo $M<1$). Per la **proprietà archimedea** (conseguenza della completezza, che vedremo con rigore nel notebook 5.2: per ogni reale esiste un naturale più grande), un tale $n$ **esiste sempre**. Quindi $1-\frac1n\in A$ supera $M$: $M$ non può essere un maggiorante.

**Conclusione**: nessun numero minore di $1$ è un maggiorante, e $1$ lo è: $\sup A=1$ (pur non essendo mai raggiunto da nessun elemento di $A$ — l'estremo superiore, a differenza del massimo, non deve necessariamente appartenere all'insieme).

**Perché questa dimostrazione a due passi (maggiorante + nessun maggiorante più piccolo) è lo schema universale per calcolare un estremo superiore:** la definizione di $\sup A$ come "minimo dei maggioranti" si traduce sempre in questi due controlli separati — mostrare che il candidato **è** un maggiorante (spesso facile), e che **nessun numero più piccolo** lo è (spesso il passo che richiede più lavoro, tipicamente usando la proprietà archimedea o un argomento simile). Questo schema tornerà identico ogni volta che dovremo calcolare un estremo superiore con rigore.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)
a_n = 1 - 1/n

# Verifica numerica: gli elementi si avvicinano a 1 restando sotto
for valore in [1, 2, 3, 10, 100, 1000]:
    print(f"n={valore}: a_n = {float(a_n.subs(n, valore)):.6f}")

# Verifica che per ogni M<1 esiste n con a_n > M
def trova_n(M):
    # 1 - 1/n > M  =>  n > 1/(1-M)
    return sp.ceiling(1/(1-M)) + 1

for M in [sp.Rational(9,10), sp.Rational(99,100), sp.Rational(999,1000)]:
    n_trovato = trova_n(M)
    valore = a_n.subs(n, n_trovato)
    print(f"\nM={M}: n={n_trovato} da  a_n={valore} > M? {valore > M}")
````

## Esercizi da risolvere

1. Trova l'estremo superiore di $A=\{x\in\mathbb{R} : x^2<9\}$, e stabilisci se appartiene ad $A$ (cioè se è anche il massimo).
2. Trova l'estremo inferiore di $B=\left\{\dfrac1n : n\in\mathbb{N},\ n\ge1\right\}$: è anche il minimo?
3. Spiega perché l'insieme $\mathbb{N}$ non ha estremo superiore in $\mathbb{R}$ (collegalo alla proprietà archimedea menzionata nell'esercizio svolto).
4. *(più impegnativo)* Dimostra (con lo schema a due passi visto nell'esercizio svolto) che $\sup\left\{2-\dfrac1{n^2} : n\ge1\right\}=2$.

<details>

**1.** $x^2<9 \iff -3<x<3$: l'insieme è $(-3,3)$. $\sup A=3$, ma $3\notin A$ (essendo $3^2=9\not<9$): non è il massimo, l'insieme non ha massimo.

**2.** $\inf B=0$ (per ogni $\varepsilon>0$ esiste $n$ con $\frac1n<\varepsilon$, proprietà archimedea). $0\notin B$ (nessun $n$ naturale dà $\frac1n=0$): non è il minimo, l'insieme non ha minimo.

**3.** Se $\mathbb{N}$ avesse un estremo superiore $M\in\mathbb{R}$, per definizione ogni naturale sarebbe $\le M$: ma questo contraddice direttamente la proprietà archimedea (per ogni reale, incluso $M$, esiste un naturale più grande). $\mathbb{N}$ non è limitato superiormente in $\mathbb{R}$.

**4.** **Maggiorante**: per ogni $n\ge1$, $\frac1{n^2}>0$, quindi $2-\frac1{n^2}<2$: sì, $2$ è un maggiorante. **Nessun maggiorante più piccolo**: dato $M<2$, cerchiamo $n$ con $2-\frac1{n^2}>M$, cioè $\frac1{n^2}<2-M$, cioè $n^2>\frac1{2-M}$, cioè $n>\frac1{\sqrt{2-M}}$ — un tale $n$ naturale esiste sempre (proprietà archimedea). Quindi nessun $M<2$ è un maggiorante: $\sup=2$.

## Riepilogo finale

- $\mathbb{R}$ si caratterizza assiomaticamente: assiomi di campo (algebra), assiomi di ordine, e l'**assioma di completezza** (ogni insieme non vuoto e limitato superiormente ha estremo superiore).
- La completezza è ciò che distingue $\mathbb{R}$ da $\mathbb{Q}$: l'insieme $\{x\in\mathbb{Q}:x^2<2\}$ non ha estremo superiore in $\mathbb{Q}$, ma lo ha in $\mathbb{R}$ (vale $\sqrt2$).
- L'**estremo superiore** è il minimo dei maggioranti; dimostrarlo richiede due passi: verificare che sia un maggiorante, e che nessun numero più piccolo lo sia.
- L'estremo superiore non deve necessariamente appartenere all'insieme (a differenza del massimo).
