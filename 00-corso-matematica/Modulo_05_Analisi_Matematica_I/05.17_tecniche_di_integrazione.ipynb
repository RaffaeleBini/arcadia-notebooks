**Prerequisiti:** 5.16 Teorema fondamentale del calcolo, 5.9 Teoremi fondamentali della derivazione, 1.3 Scomposizione dei polinomi.


## Perché ci interessa

Il teorema fondamentale del calcolo (notebook 5.16) riduce il calcolo di un integrale alla ricerca di una primitiva. Ma trovare primitive non è sempre immediato come "leggere il catalogo delle derivate al contrario" (notebook 5.9): per $x\sin x$, o $\frac1{x^2-1}$, servono tecniche apposite. Questo notebook presenta le tre più importanti — sostituzione, per parti, fratti semplici — ciascuna basata su una regola di derivazione "letta al contrario".

## Teoria, passo dopo passo

### Integrazione per sostituzione

Deriva dalla regola della catena (notebook 5.9): se $F'=f$, allora $\frac{d}{dx}F(g(x))=f(g(x))g'(x)$. Quindi:

$$\int f(g(x))g'(x)\,dx = F(g(x))+c$$

**In pratica**: poniamo $u=g(x)$, quindi $du=g'(x)\,dx$; l'integrale diventa $\int f(u)\,du=F(u)+c$, e si ritorna a $x$ sostituendo $u=g(x)$.

Esempio: $\int2x\cos(x^2)\,dx$. Poniamo $u=x^2$, $du=2x\,dx$: l'integrale diventa $\int\cos u\,du=\sin u+c=\sin(x^2)+c$.

### Integrazione per parti

Deriva dalla regola del prodotto (notebook 5.9): $(uv)'=u'v+uv'$, cioè $uv'=(uv)'-u'v$. Integrando entrambi i membri (e usando che $\int(uv)'\,dx=uv$, per il teorema fondamentale, notebook 5.16):

$$\int u(x)v'(x)\,dx = u(x)v(x) - \int u'(x)v(x)\,dx$$

**Strategia**: si sceglie $u$ come il fattore che si **semplifica** derivando (es. un polinomio, che perde grado), e $v'$ come il fattore facile da integrare.

Esempio: $\int x\sin x\,dx$. Scegliamo $u=x$ (si semplifica: $u'=1$), $v'=\sin x$ (quindi $v=-\cos x$):

$$\int x\sin x\,dx = x(-\cos x)-\int1\cdot(-\cos x)\,dx = -x\cos x+\int\cos x\,dx = -x\cos x+\sin x+c$$

### Integrazione per fratti semplici

Per funzioni razionali (notebook 3.12), si scompone il denominatore (notebook 1.3) e si riscrive la frazione come somma di pezzi più semplici, ciascuno integrabile con le tecniche note.

Esempio: $\int\dfrac1{x^2-1}\,dx$. Scomponiamo $x^2-1=(x-1)(x+1)$ e cerchiamo $A,B$ tali che $\dfrac1{(x-1)(x+1)}=\dfrac A{x-1}+\dfrac B{x+1}$. Moltiplicando per $(x-1)(x+1)$: $1=A(x+1)+B(x-1)$. Ponendo $x=1$: $1=2A\Rightarrow A=\frac12$. Ponendo $x=-1$: $1=-2B\Rightarrow B=-\frac12$. Quindi:

$$\int\frac1{x^2-1}\,dx = \frac12\int\frac1{x-1}\,dx-\frac12\int\frac1{x+1}\,dx = \frac12\ln|x-1|-\frac12\ln|x+1|+c$$

(usando che la primitiva di $\frac1{x-a}$ è $\ln|x-a|$, una lieve estensione del catalogo del notebook 5.9 al caso $x<a$, dove serve il valore assoluto, notebook 0.8, perché $\ln$ richiede argomento positivo).

````python
import sympy as sp

x, u = sp.symbols('x u', real=True)

# Verifica per sostituzione
integrale_sostituzione = sp.integrate(2*x*sp.cos(x**2), x)
print("Integrazione per sostituzione:", integrale_sostituzione)

# Verifica per parti
integrale_parti = sp.integrate(x*sp.sin(x), x)
print("Integrazione per parti:", integrale_parti)

# Verifica per fratti semplici
integrale_fratti = sp.integrate(1/(x**2-1), x)
print("Integrazione per fratti semplici:", integrale_fratti)

# Verifica esplicita della decomposizione in fratti semplici
A, B = sp.symbols('A B')
decomposizione = sp.apart(1/(x**2-1), x)
print("\nDecomposizione automatica (SymPy):", decomposizione)
````

## Esercizio svolto e commentato

**Problema:** calcola $\displaystyle\int x^2e^x\,dx$ (richiede l'integrazione per parti applicata **due volte**).

**Ragionamento:**

**Passo 1 — prima applicazione della formula per parti**, con $u=x^2$ (si semplifica derivando, $u'=2x$) e $v'=e^x$ (quindi $v=e^x$):

$$\int x^2e^x\,dx = x^2e^x - \int2xe^x\,dx$$

**Passo 2 — il nuovo integrale $\int2xe^x\,dx$ non è ancora elementare**: applichiamo di nuovo l'integrazione per parti, con $u=2x$ ($u'=2$) e $v'=e^x$ ($v=e^x$):

$$\int2xe^x\,dx = 2xe^x-\int2e^x\,dx = 2xe^x-2e^x+c$$

**Passo 3 — sostituiamo questo risultato nell'espressione del passo 1:**

$$\int x^2e^x\,dx = x^2e^x-(2xe^x-2e^x+c) = x^2e^x-2xe^x+2e^x+c'$$

**Passo 4 — raccogliamo il fattore comune $e^x$** (notebook 1.3):

$$\int x^2e^x\,dx = e^x(x^2-2x+2)+c'$$

**Verifica** (derivando il risultato, dovremmo ritrovare $x^2e^x$): $\frac{d}{dx}\big[e^x(x^2-2x+2)\big] = e^x(x^2-2x+2)+e^x(2x-2) = e^x(x^2-2x+2+2x-2) = e^x\cdot x^2 = x^2e^x$. ✓

**Perché scegliere sempre $u=$ (il polinomio) e $v'=$ (l'esponenziale) ad ogni applicazione, invece di scambiare i ruoli:** ogni volta che deriviamo il polinomio, il suo grado **diminuisce** ($x^2\to2x\to2\to0$), mentre l'esponenziale, integrato o derivato, **resta della stessa forma** ($e^x$). Scegliere sistematicamente questa direzione garantisce che il processo **termini** dopo un numero finito di passi (tanti quanto il grado del polinomio); scambiando i ruoli (derivando $e^x$ e integrando il polinomio), il grado del polinomio **aumenterebbe** a ogni passo, rendendo il problema sempre più complicato invece che più semplice.

````python
import sympy as sp

x = sp.symbols('x', real=True)
integrale = sp.integrate(x**2 * sp.exp(x), x)
print("Integrale:", sp.factor(integrale))

# Verifica derivando il risultato
verifica = sp.diff(integrale, x)
print("Derivata del risultato:", sp.simplify(verifica))
````

## Esercizi da risolvere

1. Calcola $\displaystyle\int3x^2\sin(x^3)\,dx$ (sostituzione).
2. Calcola $\displaystyle\int x\ln x\,dx$ (per parti: scegli $u=\ln x$, che si semplifica derivando).
3. Calcola $\displaystyle\int\dfrac1{x^2-4}\,dx$ (fratti semplici, stesso schema dell'esempio della teoria).
4. *(più impegnativo)* Calcola $\displaystyle\int e^x\cos x\,dx$ (per parti, applicata due volte: attenzione, dopo due applicazioni l'integrale di partenza "ricompare" — risolvi come un'equazione nell'incognita "integrale cercato").

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Con $u=x^3$, $du=3x^2dx$: $\int\sin u\,du=-\cos u+c=-\cos(x^3)+c$.

**2.** $u=\ln x$ ($u'=\frac1x$), $v'=x$ ($v=\frac{x^2}2$): $\int x\ln x\,dx=\frac{x^2}2\ln x-\int\frac{x^2}2\cdot\frac1x\,dx=\frac{x^2}2\ln x-\int\frac x2\,dx=\frac{x^2}2\ln x-\frac{x^2}4+c$.

**3.** $\frac1{x^2-4}=\frac1{(x-2)(x+2)}=\frac A{x-2}+\frac B{x+2}$. Con $x=2$: $1=4A\Rightarrow A=\frac14$. Con $x=-2$: $1=-4B\Rightarrow B=-\frac14$. Integrale: $\frac14\ln|x-2|-\frac14\ln|x+2|+c$.

**4.** Chiamiamo $I=\int e^x\cos x\,dx$. Per parti ($u=e^x$, $v'=\cos x$, $v=\sin x$): $I=e^x\sin x-\int e^x\sin x\,dx$. Il nuovo integrale, ancora per parti ($u=e^x$, $v'=\sin x$, $v=-\cos x$): $\int e^x\sin x\,dx=-e^x\cos x+\int e^x\cos x\,dx=-e^x\cos x+I$. Sostituendo: $I=e^x\sin x-(-e^x\cos x+I)=e^x\sin x+e^x\cos x-I$. Quindi $2I=e^x(\sin x+\cos x)$, cioè $I=\frac{e^x(\sin x+\cos x)}2+c$.

Verifica computazionale:

print("Esercizio 2:", sp.integrate(x*sp.log(x), x))
print("Esercizio 3:", sp.integrate(1/(x**2-4), x))
print("Esercizio 4:", sp.integrate(sp.exp(x)*sp.cos(x), x))

## Riepilogo finale

- **Sostituzione** (dalla regola della catena): $\int f(g(x))g'(x)\,dx=F(g(x))+c$, ponendo $u=g(x)$.
- **Per parti** (dalla regola del prodotto): $\int uv'\,dx=uv-\int u'v\,dx$, scegliendo $u$ come il fattore che si semplifica derivando.
- **Fratti semplici** (per funzioni razionali): scomporre il denominatore e riscrivere come somma di frazioni elementari, ciascuna con primitiva $\ln|x-a|$.
- Per integrandi con polinomio × esponenziale (o trigonometrica), l'integrazione per parti va spesso applicata più volte, con il polinomio come $u$ a ogni passo per ridurne progressivamente il grado.
