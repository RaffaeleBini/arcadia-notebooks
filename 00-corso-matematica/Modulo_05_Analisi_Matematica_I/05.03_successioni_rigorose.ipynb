## Teoria, passo dopo passo

### Dimostrazione rigorosa: unicità del limite

Nel notebook 4.3 avevamo solo intuito questo fatto. Dimostriamolo: supponiamo $a_n\to L_1$ e $a_n\to L_2$ con $L_1\ne L_2$. Sia $\varepsilon=\frac{|L_1-L_2|}2>0$ (metà della distanza tra i due candidati limite). Per definizione di limite (notebook 4.2), esiste $N_1$ tale che $|a_n-L_1|<\varepsilon$ per $n>N_1$, ed esiste $N_2$ tale che $|a_n-L_2|<\varepsilon$ per $n>N_2$. Prendendo $n>\max(N_1,N_2)$ (un indice che soddisfa entrambe le condizioni), per la disuguaglianza triangolare (notebook 0.8):

$$|L_1-L_2| \le |L_1-a_n|+|a_n-L_2| < \varepsilon+\varepsilon = |L_1-L_2|$$

Otteniamo $|L_1-L_2|<|L_1-L_2|$: una contraddizione (nessun numero è strettamente minore di sé stesso). Quindi $L_1=L_2$. ∎

**Perché la scelta $\varepsilon=\frac{|L_1-L_2|}2$ è quella "giusta":** dividendo per $2$, la somma delle due tolleranze ($\varepsilon+\varepsilon$) risulta **esattamente** $|L_1-L_2|$: è la scelta più stretta possibile che rende immediata la contraddizione. Questo tipo di scelta "a metà strada" è una tecnica standard nelle dimostrazioni ε-N.

### Dimostrazione rigorosa: algebra dei limiti (il caso della somma)

Se $a_n\to L$ e $b_n\to M$, dimostriamo $a_n+b_n\to L+M$. Fissiamo $\varepsilon>0$. Per definizione, esiste $N_1$ con $|a_n-L|<\frac\varepsilon2$ per $n>N_1$, ed esiste $N_2$ con $|b_n-M|<\frac\varepsilon2$ per $n>N_2$ (abbiamo scelto $\frac\varepsilon2$ come tolleranza per **ciascuna** delle due successioni, sapendo che sommandole ritroveremo $\varepsilon$ intero). Per $n>\max(N_1,N_2)$:

$$|(a_n+b_n)-(L+M)| = |(a_n-L)+(b_n-M)| \le |a_n-L|+|b_n-M| < \frac\varepsilon2+\frac\varepsilon2=\varepsilon$$

(disuguaglianza triangolare di nuovo). Questo è esattamente $|(a_n+b_n)-(L+M)|<\varepsilon$ per $n>\max(N_1,N_2)$: la definizione di $a_n+b_n\to L+M$. ∎

### Successioni di Cauchy: convergenza senza conoscere il limite

Un concetto nuovo, cruciale: $(a_n)$ è una **successione di Cauchy** se, per ogni $\varepsilon>0$, esiste $N$ tale che $|a_n-a_m|<\varepsilon$ per ogni $n,m>N$ (i termini si avvicinano **tra loro**, non necessariamente a un limite noto in anticipo). In $\mathbb{R}$ vale un fatto notevole (il **criterio di Cauchy**, conseguenza della completezza):

> $(a_n)$ converge $\iff$ $(a_n)$ è di Cauchy.

Questo criterio è potentissimo perché non richiede di conoscere il candidato limite: basta mostrare che i termini si avvicinano sempre di più **tra loro**. Non dimostriamo qui il criterio in dettaglio (richiede il teorema di Bolzano-Weierstrass, notebook 4.6, applicato con rigore), ma lo useremo nel notebook 5.19 per studiare la convergenza delle serie.

````python
import sympy as sp

# Verifica numerica dell'unicità del limite: se due candidati limite sono diversi,
# la stessa successione non può avvicinarsi a entrambi contemporaneamente
n = sp.symbols('n', positive=True, integer=True)
a_n = 1 + 1/n

L1, L2 = 1, 2  # due candidati, solo uno è corretto
for valore in [10, 100, 1000]:
    val = float(a_n.subs(n, valore))
    print(f"n={valore}: a_n={val:.4f}, |a_n-{L1}|={abs(val-L1):.4f}, |a_n-{L2}|={abs(val-L2):.4f}")
print("\nSolo la distanza da L1=1 si avvicina a 0: L2=2 non è il limite")

# Verifica del criterio di Cauchy su una successione convergente
def verifica_cauchy(termine, N_max, epsilon):
    valori = [float(termine.subs(n, k)) for k in range(1, N_max+1)]
    for i in range(len(valori)):
        for j in range(i+1, len(valori)):
            if abs(valori[i] - valori[j]) >= epsilon:
                return False, (i+1, j+1)
    return True, None

risultato, dettaglio = verifica_cauchy(a_n, 20, 0.1)
print(f"\nÈ di Cauchy con epsilon=0.1 sui primi 20 termini (approssimativo)? {risultato}")
````

## Esercizio svolto e commentato

**Problema:** dimostra rigorosamente (con la definizione ε-N) che se $a_n\to L$ con $L>0$, allora esiste $N$ tale che $a_n>0$ per ogni $n>N$ (un caso particolare di un risultato più generale, la **permanenza del segno**, che ritroveremo per le funzioni nel notebook 5.7).

**Ragionamento:**

**Passo 1 — scegliamo una tolleranza $\varepsilon$ opportuna.** Vogliamo mostrare che $a_n$ diventa positivo definitivamente. L'idea: se $a_n$ resta abbastanza vicino a $L>0$, non può "scendere" fino a $0$ o sotto. Scegliamo $\varepsilon=L$ (la distanza tra $L$ e $0$ stesso: la scelta più naturale, che "consuma" tutto il margine disponibile senza superarlo).

**Passo 2 — applichiamo la definizione di limite** (notebook 4.2) con questa scelta di $\varepsilon$: esiste $N$ tale che $|a_n-L|<L$ per ogni $n>N$.

**Passo 3 — "apriamo" il valore assoluto** (notebook 0.8): $|a_n-L|<L$ equivale a $-L<a_n-L<L$, cioè (sommando $L$ a tutti e tre i membri, invariantiva, notebook 0.5):

$$0 < a_n < 2L$$

**Passo 4 — leggiamo la conclusione.** In particolare, $a_n>0$ per ogni $n>N$: esattamente ciò che volevamo dimostrare.

**Perché scegliere $\varepsilon=L$ esattamente (e non un $\varepsilon$ più piccolo o più grande) è la mossa cruciale:** se avessimo scelto $\varepsilon$ più grande di $L$ (per esempio $\varepsilon=2L$), l'intervallo $(L-\varepsilon,L+\varepsilon)$ includerebbe anche numeri negativi, e la dimostrazione fallirebbe. Scegliendo $\varepsilon=L$, l'intervallo $(L-\varepsilon,L+\varepsilon)=(0,2L)$ resta **esattamente** dentro la zona positiva, senza sconfinare: la scelta della tolleranza giusta, in funzione dei dati del problema (qui, la distanza di $L$ da $0$), è un'abilità che tornerà continuamente nelle dimostrazioni ε-N e ε-δ da qui in avanti.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

# Esempio concreto: a_n = 3 - 1/n, che tende a L=3>0
a_n = 3 - 1/n
L = 3
epsilon = L  # come nella dimostrazione

# Troviamo N: |a_n - 3| < 3  =>  1/n < 3  =>  n > 1/3, quindi N=1 basta già
for valore in range(1, 6):
    val = a_n.subs(n, valore)
    print(f"n={valore}: a_n={val}, positivo? {val > 0}")
````

**1.** Scegliamo $\varepsilon=-L=|L|>0$ (essendo $L<0$). Per la definizione, esiste $N$ con $|a_n-L|<-L$ per $n>N$, cioè $L<a_n-L<-L$, cioè $2L<a_n<0$: in particolare $a_n<0$ per $n>N$.

**2.** Fissato $\varepsilon>0$, si sceglie la tolleranza $\frac\varepsilon{|c|}$ per $a_n$: esiste $N$ con $|a_n-L|<\frac\varepsilon{|c|}$ per $n>N$. Allora $|ca_n-cL|=|c||a_n-L|<|c|\cdot\frac\varepsilon{|c|}=\varepsilon$: esattamente la definizione di $ca_n\to cL$.

**3.** Perché non richiede di **conoscere in anticipo** il valore del limite (spesso difficile o impossibile da esprimere in forma chiusa per ricorsioni complesse): basta mostrare che i termini si avvicinano sempre di più tra loro, un controllo spesso più diretto da fare algebricamente sulla relazione ricorsiva stessa.

**4.** $|(-a_n)-(-L)| = |-(a_n-L)| = |a_n-L|$ (il valore assoluto di un opposto è uguale al valore assoluto originale, notebook 0.8). Quindi, dato $\varepsilon>0$, lo stesso $N$ che funziona per $a_n\to L$ (cioè $|a_n-L|<\varepsilon$ per $n>N$) funziona automaticamente anche per $-a_n\to-L$, essendo le due espressioni identiche.

## Riepilogo finale

- L'**unicità del limite** e l'**algebra dei limiti** (notebook 4.3), lasciate all'intuizione nel Modulo 4, sono ora dimostrate con piena rigorosità ε-N, tipicamente sfruttando la disuguaglianza triangolare (notebook 0.8).
- La **permanenza del segno**: se $a_n\to L$ con $L\ne0$, definitivamente $a_n$ ha lo stesso segno di $L$ — dimostrato scegliendo $\varepsilon=|L|$.
- Una **successione di Cauchy** ha termini che si avvicinano tra loro (non necessariamente a un limite noto); in $\mathbb{R}$, convergenza e proprietà di Cauchy sono equivalenti (una conseguenza della completezza).
- La scelta della tolleranza $\varepsilon$ (o sue frazioni, come $\frac\varepsilon2$) in funzione dei dati del problema è un'abilità dimostrativa centrale, che ritroveremo nei limiti di funzioni.
