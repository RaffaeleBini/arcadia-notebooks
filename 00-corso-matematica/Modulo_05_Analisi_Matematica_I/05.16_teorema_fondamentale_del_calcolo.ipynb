# 5.16 — Teorema fondamentale del calcolo

**Prerequisiti:** 5.15 Integrale di Riemann, 5.11 Teorema di Lagrange, 5.8 Derivata.


## Perché ci interessa

Derivata e integrale sembrano concetti opposti: la derivata misura una pendenza istantanea (notebook 5.8), l'integrale un'area accumulata (notebook 5.15). Il **teorema fondamentale del calcolo** rivela che sono, in un senso preciso, **operazioni inverse** — proprio come addizione e sottrazione, o esponenziale e logaritmo (notebook 3.6). Questo risultato, tra i più importanti di tutta la matematica, trasforma il calcolo di un integrale (apparentemente difficile, richiede il limite di somme) in un problema molto più semplice: trovare una funzione la cui derivata sia nota.

## Teoria, passo dopo passo

### La funzione integrale

Sia $f$ continua su $[a,b]$. Definiamo, per $x\in[a,b]$, la **funzione integrale**:

$$F(x) = \int_a^x f(t)\,dt$$

($F(x)$ è l'area accumulata da $a$ fino a $x$, che cambia al variare di $x$: è una funzione di $x$, anche se $f$ è la stessa). Usiamo $t$ come variabile "muta" di integrazione per non confonderla con l'estremo variabile $x$.

### Teorema fondamentale del calcolo, parte 1

> $F$ è derivabile su $[a,b]$, e $F'(x)=f(x)$.

**Dimostrazione**: calcoliamo il rapporto incrementale di $F$ (notebook 5.8):

$$\frac{F(x+h)-F(x)}h = \frac1h\int_x^{x+h}f(t)\,dt$$

(la differenza tra due aree accumulate è l'area del solo pezzo aggiuntivo, tra $x$ e $x+h$ — una proprietà additiva dell'integrale che non dimostriamo qui esplicitamente, ma è intuitiva). Per il **teorema della media integrale** (una conseguenza del teorema dei valori intermedi, notebook 5.7, applicato a $f$ continua): esiste $\xi$ tra $x$ e $x+h$ tale che $\int_x^{x+h}f(t)\,dt=f(\xi)\cdot h$ (l'area del pezzetto è "come se" fosse un rettangolo di altezza $f(\xi)$, per un opportuno punto intermedio). Quindi:

$$\frac{F(x+h)-F(x)}h = f(\xi)$$

Quando $h\to0$, $\xi$ (schiacciato tra $x$ e $x+h$) tende a $x$; per continuità di $f$ (notebook 5.6), $f(\xi)\to f(x)$. Quindi $F'(x)=\lim_{h\to0}\frac{F(x+h)-F(x)}h=f(x)$. ∎

### Primitive

Una funzione $G$ tale che $G'(x)=f(x)$ si chiama **primitiva** (o antiderivata) di $f$. La parte 1 del teorema dice che $F$ è **una** primitiva di $f$. Per il corollario del teorema di Lagrange (notebook 5.11: derivate uguali implicano funzioni che differiscono per una costante), **tutte** le primitive di $f$ hanno la forma $G(x)=F(x)+c$, per qualche costante $c$.

### Teorema fondamentale del calcolo, parte 2

> Se $G$ è una qualunque primitiva di $f$ (continua) su $[a,b]$, allora:
> $$\int_a^bf(x)\,dx = G(b)-G(a)$$

**Dimostrazione**: sia $F(x)=\int_a^xf(t)\,dt$ (la funzione integrale, con $F'=f$ per la parte 1). Essendo $G$ un'altra primitiva di $f$, $G(x)=F(x)+c$ per qualche costante $c$ (dalla teoria appena vista). Allora:

$$G(b)-G(a) = (F(b)+c)-(F(a)+c) = F(b)-F(a) = \int_a^bf(t)\,dt - \int_a^af(t)\,dt = \int_a^bf(t)\,dt - 0$$

(essendo $\int_a^af(t)\,dt=0$, un intervallo di larghezza nulla). ∎ Questo trasforma **radicalmente** il calcolo pratico degli integrali: invece di limiti di somme (notebook 5.15), basta trovare **una qualunque** primitiva e valutarla agli estremi — la notazione standard è $\big[G(x)\big]_a^b=G(b)-G(a)$.

````python
import sympy as sp

x, t = sp.symbols('x t', real=True)

# Verifica della parte 1: F(x) = integrale da 0 a x di t^2 dt, F'(x) dovrebbe essere x^2
F = sp.integrate(t**2, (t, 0, x))
print("F(x) =", F)
print("F'(x) =", sp.diff(F, x), " (dovrebbe coincidere con f(x)=x^2)")

# Verifica della parte 2: calcolo di un integrale definito tramite una primitiva
f = x**2
G = sp.integrate(f, x)  # primitiva generica (senza costante)
print("\nPrimitiva G(x) =", G)
integrale_definito = G.subs(x, 1) - G.subs(x, 0)
print("G(1) - G(0) =", integrale_definito)
print("Confronto con integrate diretto:", sp.integrate(f, (x, 0, 1)))
````

## Esercizio svolto e commentato

**Problema:** calcola $\int_0^\pi\sin x\,dx$ usando il teorema fondamentale del calcolo, e interpreta il risultato geometricamente.

**Ragionamento:**

**Passo 1 — troviamo una primitiva di $\sin x$.** Dal catalogo delle derivate (notebook 5.9): $\frac{d}{dx}(-\cos x)=-(-\sin x)=\sin x$. Quindi $G(x)=-\cos x$ è una primitiva di $\sin x$.

**Passo 2 — applichiamo la parte 2 del teorema:**

$$\int_0^\pi\sin x\,dx = \big[-\cos x\big]_0^\pi = (-\cos\pi)-(-\cos0) = -(-1)-(-1) = 1+1 = 2$$

**Passo 3 — interpretiamo geometricamente.** Su $[0,\pi]$, $\sin x\ge0$ (notebook 2.9: il seno è non negativo nel primo e secondo quadrante, cioè per angoli tra $0$ e $\pi$): l'integrale rappresenta davvero un'**area** (non un'area "con segno" mista), precisamente l'area sotto un arco completo della curva del seno.

**Verifica di plausibilità**: l'arco di seno su $[0,\pi]$ sta dentro il rettangolo $[0,\pi]\times[0,1]$, di area $\pi\approx3.14$; il valore trovato, $2$, è ragionevolmente minore di quest'area (l'arco non riempie tutto il rettangolo), ma non troppo piccolo (l'arco occupa una porzione consistente) — un controllo di sensatezza sempre utile dopo aver calcolato un integrale.

**Perché scegliere $G(x)=-\cos x$ (e non, per esempio, $-\cos x+5$) non cambia il risultato finale:** come visto nella teoria, tutte le primitive di $\sin x$ differiscono per una costante additiva. Nella differenza $G(b)-G(a)$, questa costante **si cancella sempre**: $(-\cos\pi+5)-(-\cos0+5)=(-\cos\pi)-(-\cos0)$, esattamente come prima. Per questo, nel calcolo di un integrale definito, si può scegliere **la primitiva più semplice possibile** (di solito omettendo del tutto la costante arbitraria), senza preoccuparsi di quale tra le infinite primitive si stia usando.

````python
import sympy as sp

x = sp.symbols('x', real=True)
G = -sp.cos(x)

integrale = G.subs(x, sp.pi) - G.subs(x, 0)
print("Integrale calcolato:", integrale)

# Verifica diretta
print("Verifica con sp.integrate:", sp.integrate(sp.sin(x), (x, 0, sp.pi)))
````

## Riepilogo finale

- **Parte 1**: la funzione integrale $F(x)=\int_a^xf(t)\,dt$ è derivabile, con $F'(x)=f(x)$: l'integrale "disfa" la derivata.
- **Primitive**: due primitive della stessa funzione differiscono sempre per una costante (conseguenza del teorema di Lagrange, notebook 5.11).
- **Parte 2**: $\int_a^bf(x)\,dx=G(b)-G(a)$, per una qualunque primitiva $G$ di $f$: trasforma il calcolo di un integrale (limiti di somme, notebook 5.15) in un problema di ricerca di primitive.
- Nel calcolo pratico, si sceglie sempre la primitiva più semplice: la costante arbitraria si cancella comunque nella differenza $G(b)-G(a)$.
