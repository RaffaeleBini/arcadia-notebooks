# 5.15 — Integrale di Riemann

**Prerequisiti:** 5.7 Teoremi fondamentali sulle funzioni continue, 4.7 Serie numeriche, 3.10 Funzioni limitate.


## Perché ci interessa

Come calcoliamo l'area sotto il grafico di una funzione? Per un rettangolo, è immediato (base per altezza); per una regione con un bordo curvo (come sotto una parabola), serve un'idea nuova. La costruzione di Riemann è elegante: **approssimare** l'area con tanti rettangoli sottili, sempre più numerosi e sempre più stretti, e prendere il limite. È lo stesso spirito delle somme parziali di una serie (notebook 4.7): un processo infinito costruito a partire da somme finite, sempre calcolabili.

## Teoria, passo dopo passo

### Partizioni e somme di Riemann

Sia $f$ limitata (notebook 3.10) su $[a,b]$. Una **partizione** di $[a,b]$ è una scelta di punti $a=x_0<x_1<\dots<x_n=b$, che divide l'intervallo in $n$ sotto-intervalli $[x_{i-1},x_i]$. Su ciascun sotto-intervallo, definiamo:

$$M_i = \sup_{[x_{i-1},x_i]} f \qquad m_i = \inf_{[x_{i-1},x_i]} f$$

(l'estremo superiore e inferiore, notebook 5.1, garantiti esistere dalla limitatezza di $f$ e dalla completezza di $\mathbb{R}$). La **somma superiore** e la **somma inferiore** sono:

$$S(P) = \sum_{i=1}^n M_i(x_i-x_{i-1}) \qquad s(P) = \sum_{i=1}^n m_i(x_i-x_{i-1})$$

Geometricamente: $S(P)$ è l'area di rettangoli che **contengono** il grafico di $f$ su ogni sotto-intervallo (sempre una sovrastima dell'area vera, se $f\ge0$); $s(P)$ è l'area di rettangoli **contenuti sotto** il grafico (sempre una sottostima). Vale sempre $s(P)\le S(P)$.

### Raffinare la partizione

Se aggiungiamo punti a una partizione (rendendola "più fine"), la somma superiore può solo **diminuire** (o restare uguale) e la somma inferiore può solo **aumentare**: rettangoli più stretti aderiscono meglio al grafico. Questo suggerisce di definire:

$$\overline{\int_a^b}f = \inf_P S(P) \qquad \underline{\int_a^b}f = \sup_P s(P)$$

(l'estremo inferiore delle somme superiori, e l'estremo superiore delle somme inferiori, su **tutte** le partizioni possibili — di nuovo garantiti dalla completezza). Vale sempre $\underline{\int}\le\overline{\int}$.

### Integrabilità secondo Riemann

$f$ è **integrabile secondo Riemann** su $[a,b]$ se $\underline{\int_a^b}f=\overline{\int_a^b}f$ (i due valori coincidono): in tal caso, questo valore comune si chiama **integrale definito**, indicato $\int_a^b f(x)\,dx$.

**Teorema (che non dimostriamo per esteso)**: ogni funzione **continua** su $[a,b]$ è integrabile secondo Riemann. L'idea: la continuità (uniforme, un rafforzamento che vedremo con più dettaglio in corsi avanzati, ma discendente dal teorema di Weierstrass, notebook 5.7, su un intervallo chiuso e limitato) garantisce che, raffinando abbastanza la partizione, $M_i-m_i$ diventi piccolo quanto si vuole su ogni sotto-intervallo, facendo collassare la differenza $S(P)-s(P)$ a $0$.

### Interpretazione come area (con segno)

Se $f\ge0$ su $[a,b]$, $\int_a^bf(x)\,dx$ è esattamente l'area della regione tra il grafico e l'asse $x$. Se $f<0$ in parte del dominio, quella parte contribuisce **negativamente**: l'integrale misura un'area "con segno".

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Costruiamo somme superiori e inferiori per f(x) = x^2 su [0,1], con partizioni sempre più fini
def somme_riemann(f_num, a, b, n):
    punti = np.linspace(a, b, n+1)
    larghezza = (b - a) / n
    somma_sup, somma_inf = 0, 0
    for i in range(n):
        x_vals_locali = np.linspace(punti[i], punti[i+1], 20)
        valori = f_num(x_vals_locali)
        somma_sup += max(valori) * larghezza
        somma_inf += min(valori) * larghezza
    return somma_sup, somma_inf

f_num = lambda t: t**2
for n in [4, 10, 100, 1000]:
    S, s = somme_riemann(f_num, 0, 1, n)
    print(f"n={n}: somma superiore={S:.6f}, somma inferiore={s:.6f}, differenza={S-s:.6f}")

integrale_esatto = sp.integrate(x**2, (x, 0, 1))
print(f"\nIntegrale esatto: {integrale_esatto}")

# Visualizzazione con rettangoli
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, n in zip(axes, [5, 20]):
    x_vals = np.linspace(0, 1, 300)
    ax.plot(x_vals, x_vals**2, color='black')
    punti = np.linspace(0, 1, n+1)
    for i in range(n):
        altezza = max(f_num(np.linspace(punti[i], punti[i+1], 10)))
        ax.bar(punti[i], altezza, width=punti[i+1]-punti[i], align='edge', alpha=0.3, edgecolor='blue')
    ax.set_title(f"Somma superiore, n={n} rettangoli")
plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** calcola $\int_0^1x\,dx$ direttamente dalla definizione, usando una partizione uniforme e il limite delle somme.

**Ragionamento:**

**Passo 1 — costruiamo una partizione uniforme** di $[0,1]$ in $n$ sotto-intervalli: $x_i=\frac in$ (per $i=0,1,\dots,n$), ciascuno di larghezza $\frac1n$.

**Passo 2 — poiché $f(x)=x$ è crescente**, su ogni sotto-intervallo $[x_{i-1},x_i]$ il minimo è $m_i=x_{i-1}=\frac{i-1}n$ e il massimo è $M_i=x_i=\frac in$ (notebook 3.9).

**Passo 3 — calcoliamo la somma superiore:**

$$S(P) = \sum_{i=1}^n \frac in\cdot\frac1n = \frac1{n^2}\sum_{i=1}^n i$$

Usando la formula della somma dei primi $n$ naturali (dimostrata per induzione nel notebook 0.20): $\sum_{i=1}^ni=\frac{n(n+1)}2$:

$$S(P) = \frac1{n^2}\cdot\frac{n(n+1)}2 = \frac{n+1}{2n}$$

**Passo 4 — calcoliamo il limite per $n\to\infty$** (raffinando la partizione, teoria):

$$\lim_{n\to\infty}\frac{n+1}{2n} = \lim_{n\to\infty}\frac12\left(1+\frac1n\right) = \frac12$$

(tecnica del raccoglimento, notebook 4.3).

**Passo 5 — verifichiamo con la somma inferiore** (dovrebbe dare lo stesso limite, se $f$ è integrabile): $s(P)=\sum_{i=1}^n\frac{i-1}n\cdot\frac1n=\frac1{n^2}\sum_{i=1}^n(i-1)=\frac1{n^2}\cdot\frac{(n-1)n}2=\frac{n-1}{2n}\to\frac12$: stesso limite. Confermato: $\int_0^1x\,dx=\frac12$.

**Perché questo valore coincide esattamente con l'area geometrica del triangolo sotto la retta $y=x$ tra $0$ e $1$:** l'integrale, per una funzione non negativa, misura l'area sotto il grafico (teoria). Qui la regione è un triangolo rettangolo con cateti di lunghezza $1$: la sua area, con la formula elementare $\frac{\text{base}\times\text{altezza}}2$, è $\frac{1\times1}2=\frac12$ — coincide esattamente con il calcolo tramite le somme di Riemann, una bella conferma che la definizione rigorosa "cattura" davvero l'intuizione geometrica di partenza.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

somma_superiore = sp.Rational(1, 1) * sp.Sum(sp.Symbol('i'), (sp.Symbol('i'), 1, n)).doit() / n**2
print("Somma superiore S(P):", sp.simplify(somma_superiore))

limite = sp.limit(somma_superiore, n, sp.oo)
print("Limite per n->infinito:", limite)

x = sp.symbols('x')
integrale_sympy = sp.integrate(x, (x, 0, 1))
print("\nIntegrale con sp.integrate:", integrale_sympy)
````

## Riepilogo finale

- L'**integrale di Riemann** si costruisce approssimando l'area sotto il grafico con somme di rettangoli (somme superiori e inferiori), raffinando la partizione fino a un limite comune.
- $f$ è integrabile se $\overline{\int}=\underline{\int}$; ogni funzione continua su $[a,b]$ è integrabile.
- L'integrale $\int_a^bf(x)\,dx$ rappresenta l'area con segno tra il grafico e l'asse $x$.
- La costruzione richiede la limitatezza di $f$: funzioni illimitate (come $\frac1x$ vicino a $0$) richiedono un'estensione, gli integrali impropri (notebook 5.18).
