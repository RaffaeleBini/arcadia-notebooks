# 5.8 — Derivata

**Prerequisiti:** 5.5 Limiti notevoli, 5.6 Continuità, 2.2 Retta.


## Perché ci interessa

Nel notebook 2.2, il coefficiente angolare $m$ di una retta misurava un tasso di variazione **costante**: quanto sale $y$ per ogni unità di $x$. Ma per una curva qualsiasi (una parabola, un'esponenziale), la "pendenza" cambia da punto a punto — pensiamo alla velocità istantanea di un'auto, diversa in ogni istante anche se la strada è la stessa. La **derivata** è lo strumento che misura questa pendenza istantanea, ed è definita — non a caso — come un limite (notebook 5.4): il concetto centrale di questo modulo, da cui deriva (letteralmente) tutto il resto.

## Teoria, passo dopo passo

### Il rapporto incrementale

Dati un punto $x_0$ e un incremento $h\ne0$, il **rapporto incrementale** di $f$ tra $x_0$ e $x_0+h$ è:

$$\frac{f(x_0+h)-f(x_0)}{h}$$

Geometricamente, è il coefficiente angolare (notebook 2.2) della retta **secante** al grafico di $f$, passante per i punti $(x_0,f(x_0))$ e $(x_0+h,f(x_0+h))$: applicando la formula del notebook 2.2, $m=\frac{\Delta y}{\Delta x}=\frac{f(x_0+h)-f(x_0)}{(x_0+h)-x_0}=\frac{f(x_0+h)-f(x_0)}h$.

### La derivata come limite del rapporto incrementale

Se il rapporto incrementale ha un limite finito per $h\to0$, questo limite si chiama **derivata di $f$ in $x_0$**:

$$f'(x_0) = \lim_{h\to0}\frac{f(x_0+h)-f(x_0)}h$$

Quando $h\to0$, il secondo punto $(x_0+h,f(x_0+h))$ si avvicina al primo lungo il grafico, e la retta secante "ruota" fino a diventare la retta **tangente** in $(x_0,f(x_0))$: $f'(x_0)$ è precisamente il coefficiente angolare di questa tangente. Se il limite non esiste (o è infinito), $f$ non è **derivabile** in $x_0$.

### Calcolo diretto: la derivata di $x^2$

Applichiamo la definizione a $f(x)=x^2$, in un punto $x_0$ generico:

$$f'(x_0) = \lim_{h\to0}\frac{(x_0+h)^2-x_0^2}h = \lim_{h\to0}\frac{x_0^2+2x_0h+h^2-x_0^2}h$$

(sviluppando il quadrato del binomio, notebook 1.2). Semplificando $x_0^2$ (si cancella) e raccogliendo $h$ al numeratore:

$$= \lim_{h\to0}\frac{2x_0h+h^2}h = \lim_{h\to0}\frac{h(2x_0+h)}h = \lim_{h\to0}(2x_0+h)$$

(semplificando $h$, lecito perché $h\ne0$ nel limite, notebook 5.4). Questo limite si calcola per sostituzione diretta (continuità del polinomio $2x_0+h$ in $h$, notebook 5.6): $f'(x_0)=2x_0$. Quindi la derivata di $x^2$ è $2x$ — una funzione essa stessa, che assegna a ogni punto la pendenza della tangente lì.

### Derivabilità implica continuità (ma non viceversa)

> Se $f$ è derivabile in $x_0$, allora $f$ è continua in $x_0$.

**Dimostrazione**: scriviamo $f(x_0+h)-f(x_0) = \dfrac{f(x_0+h)-f(x_0)}h\times h$ (moltiplicando e dividendo per $h$, lecito per $h\ne0$). Per $h\to0$: il primo fattore tende a $f'(x_0)$ (finito, per ipotesi di derivabilità), il secondo tende a $0$; per l'algebra dei limiti (notebook 5.3), il prodotto tende a $f'(x_0)\times0=0$. Quindi $f(x_0+h)\to f(x_0)$, cioè $\lim_{x\to x_0}f(x)=f(x_0)$: la definizione di continuità (notebook 5.6). ∎

**Il viceversa è falso**: $f(x)=|x|$ (notebook 0.8) è continua in $x_0=0$, ma **non derivabile** lì: il rapporto incrementale vale $\frac{|h|}h$, che è $1$ per $h>0$ e $-1$ per $h<0$ (esattamente la funzione del notebook 5.4 che non ha limite bilatero) — il grafico ha un "angolo" in $0$, senza un'unica tangente ben definita.

````python
import sympy as sp

x, h, x0 = sp.symbols('x h x0', real=True)

# Calcolo della derivata di x^2 con la definizione (rapporto incrementale)
f = x**2
rapporto_incrementale = (f.subs(x, x0 + h) - f.subs(x, x0)) / h
rapporto_semplificato = sp.simplify(rapporto_incrementale)
print("Rapporto incrementale semplificato:", rapporto_semplificato)

derivata = sp.limit(rapporto_incrementale, h, 0)
print("Derivata (limite per h->0):", derivata)

# Confronto con sp.diff (la derivata "automatica" di SymPy)
print("Derivata con sp.diff:", sp.diff(f, x))

# Verifica che |x| non sia derivabile in 0
f_abs = sp.Abs(x)
rapporto_abs = (f_abs.subs(x, 0 + h) - f_abs.subs(x, 0)) / h
lim_destro = sp.limit(rapporto_abs, h, 0, '+')
lim_sinistro = sp.limit(rapporto_abs, h, 0, '-')
print(f"\n|x| in x=0: rapporto incrementale, limite destro={lim_destro}, sinistro={lim_sinistro}")
print("Derivabile in 0?", lim_destro == lim_sinistro)
````

## Esercizio svolto e commentato

**Problema:** calcola, con la definizione, la derivata di $f(x)=\sqrt x$ in un punto generico $x_0>0$.

**Ragionamento:**

**Passo 1 — impostiamo il rapporto incrementale:**

$$\frac{f(x_0+h)-f(x_0)}h = \frac{\sqrt{x_0+h}-\sqrt{x_0}}h$$

**Passo 2 — riconosciamo la forma indeterminata "0/0"** per $h\to0$ (sia numeratore sia denominatore tendono a $0$): serve una manipolazione algebrica, come già visto per limiti simili (notebook 4.3, esercizio 4; notebook 5.4, esercizio 4).

**Passo 3 — razionalizziamo**, moltiplicando numeratore e denominatore per il "coniugato" $\sqrt{x_0+h}+\sqrt{x_0}$:

$$\frac{(\sqrt{x_0+h}-\sqrt{x_0})(\sqrt{x_0+h}+\sqrt{x_0})}{h(\sqrt{x_0+h}+\sqrt{x_0})} = \frac{(x_0+h)-x_0}{h(\sqrt{x_0+h}+\sqrt{x_0})}$$

(usando la differenza di quadrati, notebook 1.2, al numeratore).

**Passo 4 — semplifichiamo**: il numeratore diventa $h$, che si cancella con l'$h$ al denominatore (lecito, $h\ne0$):

$$\frac1{\sqrt{x_0+h}+\sqrt{x_0}}$$

**Passo 5 — calcoliamo il limite per $h\to0$.** Per continuità della radice quadrata (notebook 5.6), $\sqrt{x_0+h}\to\sqrt{x_0}$:

$$f'(x_0) = \frac1{\sqrt{x_0}+\sqrt{x_0}} = \frac1{2\sqrt{x_0}}$$

**Conclusione**: la derivata di $\sqrt x$ è $\dfrac1{2\sqrt x}$.

**Perché la razionalizzazione (moltiplicare per il "coniugato") è la tecnica standard per rapporti incrementali che coinvolgono radici:** la stessa forma indeterminata "0/0" che abbiamo incontrato più volte (notebook 4.3, esercizio 4; notebook 5.4, esercizio 4) si ripresenta qui, e la tecnica risolutiva è identica: sfruttare il prodotto notevole $(\sqrt a-\sqrt b)(\sqrt a+\sqrt b)=a-b$ per "eliminare" le radici dal numeratore, trasformando una sottrazione di radici (difficile da semplificare direttamente) in una sottrazione di quantità senza radici (qui, $x_0+h-x_0=h$), che si semplifica immediatamente con il denominatore.

````python
import sympy as sp

x0, h = sp.symbols('x0 h', positive=True)
f = sp.sqrt(x0 + h)
f0 = sp.sqrt(x0)

rapporto = (f - f0) / h
derivata = sp.limit(rapporto, h, 0)
print("Derivata di sqrt(x) in x0:", derivata)

# Confronto con sp.diff
x = sp.symbols('x', positive=True)
print("Derivata con sp.diff:", sp.diff(sp.sqrt(x), x))
````

## Riepilogo finale

- $f'(x_0)=\lim_{h\to0}\dfrac{f(x_0+h)-f(x_0)}h$: la derivata è il limite del rapporto incrementale, e rappresenta il coefficiente angolare della retta tangente al grafico in $(x_0,f(x_0))$.
- Derivabilità $\Rightarrow$ continuità, ma non viceversa: $|x|$ è continua ma non derivabile in $0$ (un "angolo" nel grafico).
- Il calcolo diretto con la definizione richiede spesso tecniche algebriche già note (sviluppo di potenze, razionalizzazione) per risolvere la forma indeterminata $\frac00$.
- La derivata di $x^2$ è $2x$; di $x^3$ è $3x^2$; di $\sqrt x$ è $\frac1{2\sqrt x}$; di $\frac1x$ è $-\frac1{x^2}$ — i primi mattoni di un catalogo che completeremo nel prossimo notebook con le regole generali di derivazione.
