## Perché ci interessa

Calcolare ogni derivata con la definizione (notebook 5.8) è possibile ma laborioso — abbiamo visto che anche $\sqrt x$ richiede una razionalizzazione non banale. Le **regole di derivazione** permettono di calcolare la derivata di funzioni complesse **combinando** le derivate di pezzi più semplici, senza mai tornare al limite del rapporto incrementale. Sono lo strumento pratico che useremo per il resto del corso.

## Teoria, passo dopo passo

### Regola della somma e del prodotto per una costante

Se $f,g$ sono derivabili in $x_0$: $(f+g)'(x_0)=f'(x_0)+g'(x_0)$, e $(cf)'(x_0)=c\cdot f'(x_0)$ per una costante $c$. Entrambe seguono direttamente dall'algebra dei limiti (notebook 5.3) applicata al rapporto incrementale della somma.

### Regola del prodotto

$$(f\cdot g)'(x_0) = f'(x_0)g(x_0)+f(x_0)g'(x_0)$$

**Dimostrazione**: il rapporto incrementale di $fg$ è $\dfrac{f(x_0+h)g(x_0+h)-f(x_0)g(x_0)}h$. Sommiamo e sottraiamo $f(x_0)g(x_0+h)$ al numeratore (un trucco algebrico, la stessa idea del "sommare zero" già vista in dimostrazioni precedenti):

$$\frac{f(x_0+h)g(x_0+h)-f(x_0)g(x_0+h)+f(x_0)g(x_0+h)-f(x_0)g(x_0)}h$$

$$= \frac{f(x_0+h)-f(x_0)}h\cdot g(x_0+h) + f(x_0)\cdot\frac{g(x_0+h)-g(x_0)}h$$

Per $h\to0$: il primo pezzo tende a $f'(x_0)\cdot g(x_0)$ (usando anche la continuità di $g$, garantita dalla sua derivabilità, notebook 5.8: $g(x_0+h)\to g(x_0)$); il secondo tende a $f(x_0)\cdot g'(x_0)$. Sommando: la formula annunciata. ∎

### Regola del quoziente

$$\left(\frac fg\right)'(x_0) = \frac{f'(x_0)g(x_0)-f(x_0)g'(x_0)}{g(x_0)^2} \qquad (g(x_0)\ne0)$$

(si dimostra in modo simile, o applicando la regola del prodotto e della composizione a $f\cdot g^{-1}$).

### Regola della catena (composizione)

Se $g$ è derivabile in $x_0$ e $f$ è derivabile in $g(x_0)$, allora $f\circ g$ (notebook 3.7) è derivabile in $x_0$:

$$(f\circ g)'(x_0) = f'(g(x_0))\cdot g'(x_0)$$

**Intuizione**: se $g$ cambia $x_0$ con "velocità" $g'(x_0)$, e $f$ cambia il suo input con "velocità" $f'(g(x_0))$ nel punto corrispondente, la variazione totale è il **prodotto** delle due velocità — una catena di effetti moltiplicati, da cui il nome.

### Derivata della funzione inversa

Se $f$ è biiettiva e derivabile, con $f'(x_0)\ne0$, e $f^{-1}$ (notebook 3.6) è continua, allora:

$$(f^{-1})'(f(x_0)) = \frac1{f'(x_0)}$$

**Intuizione**: il grafico di $f^{-1}$ è il simmetrico di quello di $f$ rispetto a $y=x$ (notebook 3.6); riflettere una retta tangente rispetto a $y=x$ **inverte** il suo coefficiente angolare (uno scambio $\Delta x\leftrightarrow\Delta y$, notebook 2.2). Questa regola è quella che useremo per trovare la derivata del logaritmo a partire da quella dell'esponenziale.

### Il catalogo delle derivate elementari

Applicando queste regole (e la definizione, per i casi base) si ottiene il catalogo che useremo sempre:

$$\frac{d}{dx}x^n=nx^{n-1} \qquad \frac{d}{dx}\sin x=\cos x \qquad \frac{d}{dx}\cos x=-\sin x \qquad \frac{d}{dx}e^x=e^x \qquad \frac{d}{dx}\ln x=\frac1x$$

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Verifica della regola del prodotto
f, g = sp.Function('f')(x), sp.Function('g')(x)
prodotto = sp.diff(f * g, x)
print("Derivata del prodotto (SymPy):", prodotto)

# Verifica della regola della catena su un esempio concreto: f(g(x)) = sin(x^2)
composizione = sp.sin(x**2)
derivata_composizione = sp.diff(composizione, x)
print("\nDerivata di sin(x^2):", derivata_composizione)
print("Verifica manuale: cos(x^2) * 2x =", sp.cos(x**2) * 2*x)
print("Coincidono?", sp.simplify(derivata_composizione - sp.cos(x**2)*2*x) == 0)

# Verifica della derivata della funzione inversa: ln(x) come inversa di e^x
# f(x) = e^x, f'(x) = e^x. (f^-1)'(y) = 1/f'(x) dove y=e^x, cioè x=ln(y)
y = sp.symbols('y', positive=True)
derivata_log = 1 / sp.exp(sp.log(y))
print("\n(f^-1)'(y) = 1/e^(ln y) =", sp.simplify(derivata_log), " (deve essere 1/y)")
print("Confronto con sp.diff(ln(y)):", sp.diff(sp.log(y), y))
````

## Esercizio svolto e commentato

**Problema:** calcola la derivata di $f(x)=x^2\sin x$ e di $g(x)=\dfrac{e^x}{x^2+1}$, usando le regole del prodotto e del quoziente.

**Ragionamento:**

**Per $f(x)=x^2\sin x$**, applichiamo la regola del prodotto (teoria) con "primo fattore" $x^2$ e "secondo fattore" $\sin x$:

$$f'(x) = \underbrace{(x^2)'}_{2x}\sin x + x^2\underbrace{(\sin x)'}_{\cos x} = 2x\sin x+x^2\cos x$$

**Per $g(x)=\dfrac{e^x}{x^2+1}$**, applichiamo la regola del quoziente, con numeratore $e^x$ (derivata $e^x$, dal catalogo) e denominatore $x^2+1$ (derivata $2x$):

$$g'(x) = \frac{(e^x)'(x^2+1)-e^x(x^2+1)'}{(x^2+1)^2} = \frac{e^x(x^2+1)-e^x\cdot2x}{(x^2+1)^2}$$

**Raccogliamo $e^x$ al numeratore** (fattore comune, notebook 1.3):

$$g'(x) = \frac{e^x(x^2+1-2x)}{(x^2+1)^2} = \frac{e^x(x-1)^2}{(x^2+1)^2}$$

(riconoscendo $x^2-2x+1=(x-1)^2$, un prodotto notevole, notebook 1.2).

**Verifica del segno**: $e^x>0$ sempre, $(x-1)^2\ge0$ sempre, $(x^2+1)^2>0$ sempre: quindi $g'(x)\ge0$ per ogni $x$, con $g'(x)=0$ solo in $x=1$. Questo ci dice, in anticipo (lo formalizzeremo studiando la monotonia tramite la derivata, prossimi notebook), che $g$ è **crescente** su tutto $\mathbb{R}$ (con un punto di "pendenza zero" in $x=1$, senza però invertire direzione).

**Perché raccogliere il fattore comune dopo aver applicato la regola del quoziente è quasi sempre un passo necessario, non opzionale:** la regola del quoziente, applicata "meccanicamente", produce spesso espressioni con fattori ripetuti (qui, $e^x$ in entrambi i termini del numeratore). Riconoscere e raccogliere questi fattori comuni (con le tecniche di scomposizione del notebook 1.3) non è solo una questione estetica: come appena visto, la forma semplificata rivela informazioni (qui, il segno di $g'$) che la forma "grezza" nasconde.

````python
import sympy as sp

x = sp.symbols('x', real=True)

f = x**2 * sp.sin(x)
derivata_f = sp.diff(f, x)
print("f'(x) =", derivata_f)

g = sp.exp(x) / (x**2 + 1)
derivata_g = sp.diff(g, x)
derivata_g_semplificata = sp.simplify(derivata_g)
print("\ng'(x) semplificata =", derivata_g_semplificata)
print("Fattorizzata:", sp.factor(derivata_g_semplificata))
````

## Riepilogo finale

- Regola del prodotto: $(fg)'=f'g+fg'$; regola del quoziente: $\left(\frac fg\right)'=\frac{f'g-fg'}{g^2}$; regola della catena: $(f\circ g)'=f'(g)\cdot g'$.
- La derivata della funzione inversa si ottiene "invertendo" la derivata: $(f^{-1})'(f(x_0))=\frac1{f'(x_0)}$.
- Il catalogo delle derivate elementari ($x^n$, $\sin x$, $\cos x$, $e^x$, $\ln x$) combinato con queste regole permette di derivare praticamente ogni funzione costruita da esse.
- Dopo aver applicato le regole, semplificare e raccogliere fattori comuni (notebook 1.3) è spesso essenziale per interpretare il risultato (per esempio, per dedurne il segno).
