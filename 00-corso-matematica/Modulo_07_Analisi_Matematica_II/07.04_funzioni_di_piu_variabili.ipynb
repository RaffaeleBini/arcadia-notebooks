## Teoria, passo dopo passo

### Definizione

Una **funzione di $n$ variabili reali** è una funzione $f:D\to\mathbb R$, dove $D\subseteq\mathbb R^n$ è il **dominio** (notebook 3.3). Scriviamo $f(x_1,\dots,x_n)$ o, più compattamente, $f(x)$ con $x=(x_1,\dots,x_n)$. In questo modulo ci concentreremo soprattutto su $n=2$ (funzioni $f(x,y)$, visualizzabili come superfici in $\mathbb R^3$) e $n=3$ (funzioni $f(x,y,z)$, più difficili da visualizzare direttamente ma altrettanto importanti in fisica).

### Il dominio: stesse regole del notebook 3.3, ma in più variabili

Come nel caso di una variabile, il dominio è l'insieme dei punti dove l'espressione ha senso: si esclude la divisione per zero (notebook 0.7), si impongono argomenti non negativi sotto radici pari (notebook 0.6), argomenti positivi per i logaritmi (notebook 1.12). La differenza è che ora queste condizioni definiscono un sottoinsieme di $\mathbb R^n$ — spesso descritto da disequazioni in più variabili — non più semplicemente un'unione di intervalli.

**Esempio**: $f(x,y)=\dfrac1{\sqrt{y-x^2}}$ richiede $y-x^2>0$ (radice al denominatore, quindi argomento strettamente positivo), cioè $y>x^2$: il dominio è la regione del piano **sopra** la parabola $y=x^2$ (notebook 2.5) — un insieme aperto (per il criterio del notebook 7.3: la condizione è una disuguaglianza stretta).

### Il grafico: una superficie in $\mathbb R^{n+1}$

Per $f:D\to\mathbb R$ con $D\subseteq\mathbb R^2$, il **grafico** è l'insieme $\{(x,y,f(x,y)) : (x,y)\in D\}\subset\mathbb R^3$: una **superficie**. È l'analogo diretto del grafico di una funzione di una variabile (notebook 3.2, una curva in $\mathbb R^2$), con una dimensione in più sia nel dominio sia nel grafico.

### Insiemi di livello: il modo pratico di "vedere" una superficie in 2D

Visualizzare direttamente una superficie in 3D è spesso difficile su carta. Uno strumento alternativo, molto usato (per esempio nelle mappe topografiche, con le curve di livello dell'altitudine): l'**insieme di livello** $c$ di $f$ è:

$$L_c = \{(x,y)\in D : f(x,y)=c\}$$

Per $f(x,y)=x^2+y^2$ (già vista nel notebook 7.1), gli insiemi di livello $L_c$ (per $c>0$) sono circonferenze di raggio $\sqrt c$ centrate nell'origine: disegnando $L_c$ per diversi valori di $c$ (le "curve di livello"), si ottiene un'immagine 2D che rappresenta fedelmente la forma del paraboloide 3D — più le curve sono ravvicinate, più la superficie è "ripida" in quella zona, un'intuizione che formalizzeremo con il gradiente (notebook 7.9).

### Funzioni vettoriali (anteprima)

Più in generale, si possono considerare funzioni $F:D\to\mathbb R^m$ con $D\subseteq\mathbb R^n$ (funzioni **vettoriali**, che restituiscono un vettore invece di un numero): $F(x)=(F_1(x),\dots,F_m(x))$, dove ogni **componente** $F_i:D\to\mathbb R$ è una funzione scalare come quelle appena definite. Studieremo questo caso a fondo nel notebook 7.20 (campi vettoriali); per ora, il caso $m=1$ (funzioni scalari) sarà il nostro oggetto principale.

````python
import numpy as np
import matplotlib.pyplot as plt

# Grafico e curve di livello di f(x,y) = x^2 + y^2
X = np.linspace(-3, 3, 100)
Y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(X, Y)
Z = X**2 + Y**2

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Curve di livello (contour plot)
livelli = [1, 2, 4, 6, 8]
cs = axes[0].contour(X, Y, Z, levels=livelli, cmap='viridis')
axes[0].clabel(cs, inline=True, fontsize=8)
axes[0].set_title("Curve di livello di f(x,y) = x^2+y^2")
axes[0].set_aspect('equal')

# Grafico 3D come confronto
ax3d = fig.add_subplot(1, 2, 2, projection='3d')
ax3d.plot_surface(X, Y, Z, cmap='viridis', alpha=0.7)
ax3d.set_title("Grafico 3D (superficie)")

plt.tight_layout()
plt.show()

# Dominio di f(x,y) = 1/sqrt(y-x^2)
import sympy as sp
x, y = sp.symbols('x y', real=True)
condizione = y - x**2 > 0
print("Dominio: {(x,y) : y > x^2}")
````

## Esercizio svolto e commentato

**Problema:** trova il dominio di $f(x,y)=\ln(4-x^2-y^2)+\dfrac1{x-y}$, e stabilisci se è aperto, chiuso, né l'uno né l'altro (notebook 7.3).

**Ragionamento:**

**Passo 1 — individuiamo le due condizioni indipendenti**: il logaritmo richiede argomento positivo (notebook 1.12): $4-x^2-y^2>0$, cioè $x^2+y^2<4$. La frazione richiede denominatore non nullo (notebook 0.7): $x-y\ne0$, cioè $x\ne y$.

**Passo 2 — il dominio è l'intersezione** delle due condizioni: $D=\{(x,y) : x^2+y^2<4 \text{ e } x\ne y\}$ — il disco aperto di raggio $2$, **privato** della retta $y=x$ (dove la funzione non è definita).

**Passo 3 — stabiliamo se $D$ è aperto.** Entrambe le condizioni che definiscono $D$ sono disuguaglianze **strette** (o disuguaglianze di "non appartenenza", equivalenti a un aperto: $\{x\ne y\}$ è il complementare della retta chiusa $\{x=y\}$, quindi è aperto). Per il risultato dell'esercizio 4 del notebook 7.3, l'**intersezione** di due insiemi aperti è aperta.

**Conclusione:** $D$ è **aperto** (non è chiuso: per esempio, un punto sulla circonferenza $x^2+y^2=4$ appartiene alla frontiera ma non a $D$, quindi $D\ne\overline D$).

**Perché conviene sempre scomporre il dominio in condizioni elementari indipendenti, come nel passo 1, prima di descriverlo geometricamente:** ogni condizione elementare (denominatore non nullo, radicando non negativo, argomento del logaritmo positivo) corrisponde a un tipo di insieme già familiare dal notebook 7.3 (aperto o chiuso, a seconda che la disuguaglianza sia stretta o larga); il dominio completo, essendo sempre l'**intersezione** di queste condizioni elementari (dovendo valere **tutte** contemporaneamente), eredita le proprietà topologiche di ciascuna tramite risultati generali come quello dell'esercizio 4 del notebook 7.3 — evitando di dover riverificare la definizione di aperto/chiuso da zero ogni volta.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

condizione1 = sp.Lt(x**2 + y**2, 4)  # x^2+y^2 < 4
print("Condizione 1 (logaritmo):", condizione1)
print("Condizione 2 (frazione): x != y")

# Verifica numerica: alcuni punti dentro/fuori dal dominio
punti = [(0, 0), (1, 0), (1, 1), (1.5, 1.5), (0, 1.9)]
for px, py in punti:
    cond1 = px**2 + py**2 < 4
    cond2 = px != py
    print(f"({px},{py}): x^2+y^2<4? {cond1}, x!=y? {cond2}, nel dominio? {cond1 and cond2}")
````

## Riepilogo finale

- Una **funzione di $n$ variabili** $f:D\to\mathbb R$, con $D\subseteq\mathbb R^n$, generalizza direttamente il concetto del notebook 3.1; il dominio si trova con le stesse regole del notebook 3.3, applicate in più variabili.
- Il **grafico** di $f(x,y)$ è una superficie in $\mathbb R^3$; le **curve di livello** $\{f(x,y)=c\}$ sono uno strumento pratico per visualizzarla in 2D.
- Il dominio, essendo tipicamente intersezione di condizioni elementari, eredita le proprietà topologiche (aperto/chiuso) di ciascuna condizione (notebook 7.3).
- Le **funzioni vettoriali** $F:\mathbb R^n\to\mathbb R^m$ generalizzano ulteriormente, restituendo un vettore invece di uno scalare: le studieremo a fondo più avanti nel modulo.
