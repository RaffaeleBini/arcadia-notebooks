## Perché ci interessa

Nel notebook 5.15, l'integrale $\int_a^bf(x)dx$ calcolava l'area sotto una curva, come limite di somme di rettangoli sempre più sottili. Per una funzione $f(x,y)\ge0$ di due variabili, il grafico è una superficie, e ha senso chiedersi: qual è il **volume** sotto quella superficie, sopra una regione del piano? La risposta è l'**integrale doppio**, costruito con la stessa idea — approssimare con parallelepipedi invece che rettangoli — e il fatto notevole è che, nella pratica, si calcola riducendolo a **due integrali unidimensionali successivi**, riusando tutte le tecniche già note dal Modulo 5.

## Teoria, passo dopo passo

### Costruzione dell'integrale doppio (idea)

Sia $f(x,y)\ge0$ definita su un **rettangolo** $R=[a,b]\times[c,d]=\{(x,y):a\le x\le b,\,c\le y\le d\}$. Suddividiamo $R$ in una griglia di piccoli rettangoli, esattamente come nel notebook 5.15 suddividevamo $[a,b]$ in sottointervalli. Su ogni rettangolino di area $\Delta x\,\Delta y$, approssimiamo il volume sotto il grafico con un parallelepipedo di altezza $f$ (valutata in un punto del rettangolino) e base $\Delta x\Delta y$. Sommando su tutta la griglia (una **somma di Riemann doppia**) e raffinando la griglia all'infinito (esattamente come per le somme superiori/inferiori del notebook 5.15), si ottiene — se il limite esiste — l'**integrale doppio**:

$$\iint_R f(x,y)\,dA = \lim_{\text{griglia}\to0} \sum_{i,j} f(x_i,y_j)\,\Delta x\,\Delta y$$

Come nel caso unidimensionale, ogni funzione **continua** su un rettangolo (dominio compatto, notebook 7.3, 7.6) è integrabile.

### Il teorema di Fubini: ridurre a integrali iterati

> **Teorema (Fubini).** Se $f$ è continua su $R=[a,b]\times[c,d]$, allora:
> $$\iint_R f(x,y)\,dA = \int_a^b\left(\int_c^df(x,y)\,dy\right)dx = \int_c^d\left(\int_a^bf(x,y)\,dx\right)dy$$

Questo risultato (che non dimostriamo nel dettaglio, ma la cui idea è che entrambe le forme approssimano lo stesso volume, "affettando" il solido lungo $x$ o lungo $y$) trasforma un problema bidimensionale in **due integrali unidimensionali successivi**: si integra prima rispetto a una variabile (tenendo l'altra fissa, esattamente come per le derivate parziali, notebook 7.7), poi si integra il risultato rispetto all'altra — usando tutte le tecniche del notebook 5.17 (sostituzione, per parti).

### Integrazione su regioni non rettangolari

Per una regione $D$ più generale (non un rettangolo), tipicamente descritta come $D=\{(x,y):a\le x\le b,\ \phi_1(x)\le y\le\phi_2(x)\}$ (compresa tra due curve, al variare di $x$ in un intervallo), l'integrale doppio si calcola con i **limiti di integrazione variabili**:

$$\iint_D f(x,y)\,dA = \int_a^b\left(\int_{\phi_1(x)}^{\phi_2(x)}f(x,y)\,dy\right)dx$$

L'integrale interno, rispetto a $y$, tratta $x$ come una costante nei limiti di integrazione (esattamente come lo era nell'integranda per le derivate parziali) — il risultato è una funzione di $x$ soltanto, che si integra poi normalmente.

### Area come integrale doppio

Ponendo $f(x,y)=1$, l'integrale doppio $\iint_D1\,dA$ restituisce semplicemente l'**area** di $D$: la somma delle aree dei rettangolini della griglia, che nel limite dà l'area esatta della regione.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

# Integrale doppio di f(x,y) = x*y sul rettangolo [0,1] x [0,2]
f = x * y
integrale_interno = sp.integrate(f, (y, 0, 2))
print("Integrale interno (rispetto a y):", integrale_interno)
integrale_totale = sp.integrate(integrale_interno, (x, 0, 1))
print("Integrale doppio (Fubini):", integrale_totale)

# Verifica: integrando nell'ordine opposto, si ottiene lo stesso risultato
integrale_interno2 = sp.integrate(f, (x, 0, 1))
integrale_totale2 = sp.integrate(integrale_interno2, (y, 0, 2))
print("Integrale doppio (ordine invertito):", integrale_totale2)

# Integrale su una regione non rettangolare: D = {0<=x<=1, x^2<=y<=x}
f2 = x + y
integrale_regione = sp.integrate(sp.integrate(f2, (y, x**2, x)), (x, 0, 1))
print("\nIntegrale su D={0<=x<=1, x^2<=y<=x}:", integrale_regione)

# Area della stessa regione (f=1)
area = sp.integrate(sp.integrate(1, (y, x**2, x)), (x, 0, 1))
print("Area della regione D:", area)
````

## Esercizio svolto e commentato

**Problema:** calcola $\iint_Dxy\,dA$, dove $D$ è la regione compresa tra $y=x^2$ e $y=x$, per $0\le x\le1$ (la stessa regione dell'esempio numerico sopra, ora con un integrando diverso).

**Ragionamento:**

**Passo 1 — descriviamo $D$ come regione compresa tra due curve** al variare di $x$: per $0\le x\le1$, notiamo che $x^2\le x$ in questo intervallo (verificabile: $x-x^2=x(1-x)\ge0$ per $0\le x\le1$), quindi $D=\{0\le x\le1,\ x^2\le y\le x\}$.

**Passo 2 — impostiamo l'integrale iterato**, integrando prima rispetto a $y$ (tenendo $x$ fisso):

$$\iint_Dxy\,dA = \int_0^1\left(\int_{x^2}^xxy\,dy\right)dx$$

**Passo 3 — calcoliamo l'integrale interno.** Trattando $x$ come costante: $\int_{x^2}^xxy\,dy=x\left[\dfrac{y^2}2\right]_{x^2}^x=x\left(\dfrac{x^2}2-\dfrac{x^4}2\right)=\dfrac{x^3-x^5}2$ (usando il teorema fondamentale del calcolo, notebook 5.16, sulla primitiva $y^2/2$ di $y$).

**Passo 4 — integriamo il risultato rispetto a $x$**:

$$\int_0^1\frac{x^3-x^5}2\,dx = \frac12\left[\frac{x^4}4-\frac{x^6}6\right]_0^1 = \frac12\left(\frac14-\frac16\right) = \frac12\times\frac{3-2}{12} = \frac12\times\frac1{12} = \frac1{24}$$

**Conclusione:** $\iint_Dxy\,dA=\dfrac1{24}$.

**Perché conviene sempre determinare con precisione **quale curva sta sopra e quale sotto** (passo 1), prima di impostare i limiti di integrazione:** i limiti dell'integrale interno (rispetto a $y$) vanno dal valore **più basso** al valore **più alto** che $y$ assume nella regione, per ogni $x$ fissato; scambiare per errore $x^2$ e $x$ come limiti inferiore/superiore darebbe un integrale con segno sbagliato (o comunque errato) — un controllo geometrico rapido (verificare quale delle due curve è "sopra" l'altra nell'intervallo di interesse) evita questo errore comune.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = x * y

integrale_interno = sp.integrate(f, (y, x**2, x))
print("Integrale interno:", sp.simplify(integrale_interno))

integrale_totale = sp.integrate(integrale_interno, (x, 0, 1))
print("Integrale doppio totale:", integrale_totale)
````

## Esercizi da risolvere

1. Calcola $\iint_Rx^2y\,dA$ sul rettangolo $R=[0,2]\times[0,3]$.
2. Calcola $\iint_D(x+y)\,dA$, dove $D=\{0\le x\le1,\ 0\le y\le1-x\}$ (un triangolo).
3. Usa un integrale doppio per calcolare l'area della regione compresa tra $y=x^2$ e $y=2-x^2$.
4. *(più impegnativo)* Calcola $\iint_D e^{x^2}\,dA$, dove $D=\{0\le y\le1,\ y\le x\le1\}$ (suggerimento: $e^{x^2}$ non ha primitiva elementare rispetto a $x$; prova a **scambiare l'ordine di integrazione**, integrando prima rispetto a $y$).

**1.** $\int_0^2\left(\int_0^3x^2y\,dy\right)dx=\int_0^2x^2\left[\dfrac{y^2}2\right]_0^3dx=\int_0^2x^2\times\dfrac92\,dx=\dfrac92\left[\dfrac{x^3}3\right]_0^2=\dfrac92\times\dfrac83=12$.

**2.** $\int_0^1\left(\int_0^{1-x}(x+y)\,dy\right)dx=\int_0^1\left[xy+\dfrac{y^2}2\right]_0^{1-x}dx=\int_0^1\left(x(1-x)+\dfrac{(1-x)^2}2\right)dx$. Sviluppando: $x-x^2+\dfrac{1-2x+x^2}2=x-x^2+\dfrac12-x+\dfrac{x^2}2=\dfrac12-\dfrac{x^2}2$. Integrando: $\int_0^1\left(\dfrac12-\dfrac{x^2}2\right)dx=\dfrac12-\dfrac16=\dfrac13$.

**3.** Le curve si intersecano dove $x^2=2-x^2$, cioè $x^2=1$, $x=\pm1$. Area $=\int_{-1}^1\left(\int_{x^2}^{2-x^2}1\,dy\right)dx=\int_{-1}^1(2-2x^2)dx=\left[2x-\dfrac{2x^3}3\right]_{-1}^1=\left(2-\dfrac23\right)-\left(-2+\dfrac23\right)=4-\dfrac43=\dfrac83$.

**4.** Scambiando l'ordine: $D=\{0\le y\le1,\ y\le x\le1\}$ diventa, scambiando i ruoli, $D=\{0\le x\le1,\ 0\le y\le x\}$ (stessa regione triangolare, descritta al variare di $x$ invece che di $y$). $\int_0^1\left(\int_0^xe^{x^2}dy\right)dx=\int_0^1e^{x^2}\times x\,dx$ (l'integrale interno, rispetto a $y$, di una costante $e^{x^2}$ dà semplicemente $e^{x^2}\times x$). Ora $\int_0^1xe^{x^2}dx$ si risolve per sostituzione (notebook 5.17) con $u=x^2$, $du=2x\,dx$: $\dfrac12\int_0^1e^u\,du=\dfrac12(e-1)$.

## Riepilogo finale

- L'**integrale doppio** $\iint_Rf\,dA$ generalizza l'integrale di Riemann (notebook 5.15) a due dimensioni, come limite di somme su una griglia; misura il volume sotto il grafico di $f\ge0$.
- Il **teorema di Fubini** riduce l'integrale doppio a due integrali unidimensionali successivi, calcolabili con le tecniche già note (Modulo 5).
- Su regioni non rettangolari, i limiti dell'integrale interno diventano **funzioni** dell'altra variabile, descrivendo le curve che delimitano la regione.
- Ponendo $f=1$, l'integrale doppio calcola l'**area** della regione; scambiare l'ordine di integrazione può semplificare drasticamente il calcolo (come nell'esercizio 4).
