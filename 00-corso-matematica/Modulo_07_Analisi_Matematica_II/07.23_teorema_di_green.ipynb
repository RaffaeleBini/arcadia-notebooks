# 7.23 — Teorema di Green

**Prerequisiti:** 7.22 Rotore, 7.18 Integrali curvilinei, 7.16 Integrali doppi, 5.16 Teorema fondamentale del calcolo.


## Perché ci interessa

Il teorema fondamentale del calcolo (notebook 5.16) collegava un integrale su un intervallo $[a,b]$ al valore della primitiva **sul bordo** $\{a,b\}$ (due punti). Il **teorema di Green** è l'analogo bidimensionale: collega un integrale doppio su una regione $D$ del piano a un integrale curvilineo lungo il **bordo** di $D$ (una curva chiusa). È il primo di tre grandi teoremi (Green, divergenza, Stokes) che, in questo modulo, generalizzano progressivamente il teorema fondamentale del calcolo a dimensioni superiori — tutti espressioni dello stesso principio unificante: "l'integrale di una derivata sull'interno è determinato dai valori sul bordo".

## Teoria, passo dopo passo

### Curve chiuse orientate positivamente

Una curva **chiusa** ha punto iniziale coincidente col punto finale (percorre un "anello"). Diciamo che è percorsa in verso **positivo** (o antiorario) se, muovendosi lungo la curva, la regione $D$ racchiusa resta sempre alla **sinistra** — la convenzione standard, coerente con l'orientamento antiorario delle coordinate polari (notebook 2.13).

### Enunciato del teorema

> **Teorema (Green).** Sia $D\subset\mathbb R^2$ una regione delimitata da una curva chiusa $C=\partial D$, percorsa in verso positivo. Se $\mathbf F=(P,Q)$ ha componenti di classe $C^1$ su $D$, allora:
> $$\oint_CP\,dx+Q\,dy = \iint_D\left(\frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)dA$$

(il simbolo $\oint$ indica un integrale curvilineo lungo una curva **chiusa**). Riconosciamo l'integranda a destra: è esattamente la **componente $z$ del rotore** di $\mathbf F$ (notebook 7.22)! Il teorema di Green si può quindi riscrivere come $\oint_C\mathbf F\cdot d\mathbf r=\iint_D(\operatorname{rot}\mathbf F)_z\,dA$: la circolazione lungo il bordo è uguale all'integrale della "rotazione locale" su tutta la regione — le rotazioni locali "si sommano" fino a dare la circolazione complessiva sul bordo, con le rotazioni interne che si cancellano a vicenda (un'intuizione che la dimostrazione formale rende rigorosa).

### Conseguenza: conservativo su domini semplicemente connessi

Il teorema di Green completa il collegamento lasciato in sospeso nel notebook 7.20-7.22: se $\partial P/\partial y=\partial Q/\partial x$ ovunque in un dominio **semplicemente connesso** (senza "buchi", così che ogni curva chiusa al suo interno racchiuda solo punti del dominio), allora per **ogni** curva chiusa $C$ in quel dominio, $\oint_C\mathbf F\cdot d\mathbf r=\iint_D0\,dA=0$: la circolazione è sempre nulla, il che implica (con un argomento che collega curve chiuse a percorsi aperti con gli stessi estremi) che il lavoro dipende solo dagli estremi — cioè $\mathbf F$ è conservativo. Questo rende la condizione del notebook 7.20 **sufficiente**, non solo necessaria, su domini di questo tipo.

### Applicazione: calcolo di aree

Ponendo $\mathbf F=(-y/2,\,x/2)$ (o altre scelte equivalenti, come $(0,x)$ o $(-y,0)$): $\partial Q/\partial x-\partial P/\partial y=1/2-(-1/2)=1$. Il teorema di Green dà:

$$\text{Area}(D) = \iint_D1\,dA = \oint_C\frac{-y}2dx+\frac x2dy$$

Questa formula, sorprendente a prima vista, permette di calcolare l'**area** di una regione integrando solo lungo il suo **bordo** — utile, per esempio, per calcolare aree racchiuse da curve parametriche complicate senza dover impostare un integrale doppio esplicito.

````python
import sympy as sp

t = sp.symbols('t', real=True)

# Verifica del teorema di Green per F=(P,Q)=(-y,x) sulla circonferenza unitaria
# Confrontiamo l'integrale curvilineo (già calcolato nel notebook 7.18/7.22) con l'integrale doppio
x, y = sp.symbols('x y', real=True)
P, Q = -y, x

rotore_z = sp.diff(Q, x) - sp.diff(P, y)
print("Componente z del rotore (integranda):", rotore_z)

r_polar, theta = sp.symbols('r theta', positive=True)
integrale_doppio = sp.integrate(sp.integrate(rotore_z * r_polar, (r_polar, 0, 1)), (theta, 0, 2*sp.pi))
print("Integrale doppio su D (disco unitario):", integrale_doppio)

# Integrale curvilineo diretto lungo la circonferenza
r_curva = sp.Matrix([sp.cos(t), sp.sin(t)])
r_prime = r_curva.diff(t)
F_lungo_curva = sp.Matrix([P, Q]).subs({x: r_curva[0], y: r_curva[1]})
integranda_curva = (F_lungo_curva.T * r_prime)[0]
integrale_curvilineo = sp.integrate(sp.simplify(integranda_curva), (t, 0, 2*sp.pi))
print("Integrale curvilineo lungo il bordo:", integrale_curvilineo)
print("\nCoincidono (teorema di Green verificato)?", integrale_doppio == integrale_curvilineo)
````

## Esercizio svolto e commentato

**Problema:** usa il teorema di Green per calcolare $\oint_C(x^2-y)\,dx+(x+y^2)\,dy$, dove $C$ è il bordo del quadrato $[0,1]\times[0,1]$, percorso in verso antiorario.

**Ragionamento:**

**Passo 1 — invece di calcolare direttamente l'integrale curvilineo** (che richiederebbe di parametrizzare separatamente i quattro lati del quadrato, sommando quattro contributi), **applichiamo il teorema di Green**, trasformandolo in un integrale doppio.

**Passo 2 — identifichiamo $P=x^2-y$, $Q=x+y^2$**, e calcoliamo l'integranda del teorema: $\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}=1-(-1)=2$.

**Passo 3 — l'integranda è costante** ($=2$): l'integrale doppio si riduce a $2$ volte l'area del quadrato.

**Passo 4 — calcoliamo**: $\iint_D2\,dA=2\times\text{Area}([0,1]\times[0,1])=2\times1=2$.

**Conclusione:** $\oint_C(x^2-y)dx+(x+y^2)dy=2$.

**Perché questo esempio mostra vividamente il vantaggio pratico del teorema di Green:** calcolare direttamente l'integrale curvilineo lungo i quattro lati del quadrato avrebbe richiesto quattro parametrizzazioni distinte e altrettanti integrali unidimensionali da sommare (con attenzione ai segni, dato il verso di percorrenza di ciascun lato); il teorema di Green riduce tutto a un singolo integrale doppio, qui persino banale (integranda costante) grazie alla struttura polinomiale semplice di $P,Q$ — un guadagno di efficienza tipico ogni volta che l'integranda del teorema di Green ($\partial Q/\partial x-\partial P/\partial y$) risulta più semplice delle componenti originali $P,Q$.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
P = x**2 - y
Q = x + y**2

integranda = sp.diff(Q, x) - sp.diff(P, y)
print("Integranda (rot_z):", integranda)

integrale = sp.integrate(sp.integrate(integranda, (y, 0, 1)), (x, 0, 1))
print("Integrale doppio (= integrale curvilineo per Green):", integrale)
````

## Riepilogo finale

- Il **teorema di Green** collega un integrale curvilineo lungo il bordo di una regione piana a un integrale doppio sulla regione: $\oint_CP\,dx+Q\,dy=\iint_D(\partial Q/\partial x-\partial P/\partial y)\,dA$.
- L'integranda del teorema è esattamente la **componente $z$ del rotore** (notebook 7.22): il teorema è il caso bidimensionale di un principio più generale.
- Su domini **semplicemente connessi**, il teorema rende la condizione $\partial P/\partial y=\partial Q/\partial x$ **sufficiente** (non solo necessaria) per la conservatività.
- Una scelta astuta di $\mathbf F$ (come $(-y/2,x/2)$) trasforma il teorema in una formula per calcolare **aree** tramite un integrale sul solo bordo.
