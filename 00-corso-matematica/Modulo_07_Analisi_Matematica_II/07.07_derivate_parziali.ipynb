# 7.7 — Derivate parziali

**Prerequisiti:** 7.6 Continuità, 5.8 Derivata.


## Perché ci interessa

Nel notebook 5.8, la derivata $f'(x_0)$ misurava la pendenza istantanea di una curva $y=f(x)$. Per una superficie $z=f(x,y)$, la "pendenza" non è un unico numero: dipende dalla **direzione** in cui ci si muove sulla superficie. Muovendosi verso est la superficie può salire, muovendosi verso nord può scendere. Il modo più semplice per iniziare a catturare questa informazione è studiare la pendenza lungo le direzioni degli **assi coordinati**: fissare tutte le variabili tranne una, e derivare rispetto a quella — ottenendo le **derivate parziali**.

## Teoria, passo dopo passo

### Definizione

Sia $f:D\to\mathbb R$, $D\subseteq\mathbb R^2$, e $(x_0,y_0)$ un punto interno a $D$ (notebook 7.3: esiste una palla attorno interamente in $D$). La **derivata parziale rispetto a $x$** in $(x_0,y_0)$ è:

$$\frac{\partial f}{\partial x}(x_0,y_0) = \lim_{h\to0}\frac{f(x_0+h,y_0)-f(x_0,y_0)}h$$

cioè: si tiene $y=y_0$ **fisso**, e si deriva $g(x)=f(x,y_0)$ (una funzione di una sola variabile!) rispetto a $x$, esattamente con la definizione del notebook 5.8. Analogamente, la **derivata parziale rispetto a $y$**:

$$\frac{\partial f}{\partial y}(x_0,y_0) = \lim_{h\to0}\frac{f(x_0,y_0+h)-f(x_0,y_0)}h$$

(tenendo $x=x_0$ fisso). Notazioni equivalenti: $\partial_xf$, $f_x$, $\partial f/\partial x$.

### Calcolo pratico: le stesse regole del notebook 5.9

Poiché la derivata parziale rispetto a $x$ è, per definizione, **la derivata ordinaria** della funzione di una variabile ottenuta "congelando" $y$, tutte le regole di derivazione del notebook 5.9 (somma, prodotto, quoziente, catena) si applicano **direttamente**, trattando le altre variabili come costanti.

**Esempio**: per $f(x,y)=x^2y^3+\sin(xy)$: $\dfrac{\partial f}{\partial x}=2xy^3+y\cos(xy)$ (derivando rispetto a $x$, trattando $y$ come costante: $y^3$ resta un fattore costante, e la derivata di $\sin(xy)$ rispetto a $x$ usa la regola della catena, notebook 5.9, con $y$ costante che compare come fattore). Analogamente $\dfrac{\partial f}{\partial y}=3x^2y^2+x\cos(xy)$.

### Interpretazione geometrica

$\dfrac{\partial f}{\partial x}(x_0,y_0)$ è la pendenza della retta tangente alla curva ottenuta **intersecando** il grafico di $f$ con il piano verticale $y=y_0$ — la "sezione" della superficie lungo la direzione $x$, in quel punto. Analogamente per $\partial f/\partial y$, con il piano $x=x_0$.

### Attenzione: le derivate parziali NON garantiscono la continuità

Nel caso unidimensionale, derivabilità implicava continuità (notebook 5.8). Sorprendentemente, in più variabili, **l'esistenza di entrambe le derivate parziali non implica la continuità**. Consideriamo $f(x,y)=\dfrac{xy}{x^2+y^2}$ per $(x,y)\ne(0,0)$, $f(0,0)=0$ — la funzione del notebook 7.5, che non è nemmeno continua nell'origine (il limite non esiste). Eppure:

$$\frac{\partial f}{\partial x}(0,0) = \lim_{h\to0}\frac{f(h,0)-f(0,0)}h = \lim_{h\to0}\frac{0-0}h = 0$$

(perché $f(h,0)=\frac{h\times0}{h^2+0}=0$ per ogni $h\ne0$). Analogamente $\partial f/\partial y(0,0)=0$. Entrambe le derivate parziali **esistono** (e valgono $0$), ma $f$ **non è continua** in $(0,0)$! Questo accade perché le derivate parziali guardano solo lungo gli assi — esattamente le due direzioni ($m=0$ e la retta verticale) lungo cui $f$ **si annulla**; nessuna informazione sul comportamento lungo le altre infinite direzioni. Questa "falla" motiverà il concetto più forte di **differenziabilità**, nel prossimo notebook.

````python
import sympy as sp

x, y, h = sp.symbols('x y h', real=True)

f = x**2 * y**3 + sp.sin(x * y)
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
print("df/dx =", df_dx)
print("df/dy =", df_dy)

# Verifica con la definizione del rapporto incrementale, in un punto specifico
x0, y0 = 1, 2
rapporto_x = (f.subs({x: x0 + h, y: y0}) - f.subs({x: x0, y: y0})) / h
limite_x = sp.limit(rapporto_x, h, 0)
print(f"\ndf/dx in ({x0},{y0}) via definizione:", limite_x)
print(f"df/dx in ({x0},{y0}) via sp.diff:", df_dx.subs({x: x0, y: y0}))

# Il controesempio: xy/(x^2+y^2) ha derivate parziali in (0,0) ma non è continua lì
f2 = (x * y) / (x**2 + y**2)
dfx2_in_0 = sp.limit((f2.subs({x: h, y: 0}) - 0) / h, h, 0)
dfy2_in_0 = sp.limit((f2.subs({x: 0, y: h}) - 0) / h, h, 0)
print("\nControesempio: df/dx(0,0) =", dfx2_in_0, " df/dy(0,0) =", dfy2_in_0)
print("(entrambe esistono e sono 0, ma f non è continua in (0,0) - vedi notebook 7.5)")
````

## Esercizio svolto e commentato

**Problema:** calcola $\partial f/\partial x$ e $\partial f/\partial y$ per $f(x,y)=\dfrac{x}{x^2+y^2}$, e valutale in $(1,1)$.

**Ragionamento:**

**Passo 1 — deriviamo rispetto a $x$, trattando $y$ come costante.** Usiamo la regola del quoziente (notebook 5.9): $\dfrac{\partial}{\partial x}\left(\dfrac{x}{x^2+y^2}\right)=\dfrac{1\times(x^2+y^2)-x\times2x}{(x^2+y^2)^2}=\dfrac{x^2+y^2-2x^2}{(x^2+y^2)^2}=\dfrac{y^2-x^2}{(x^2+y^2)^2}$.

**Passo 2 — deriviamo rispetto a $y$, trattando $x$ come costante.** Il numeratore $x$ non dipende da $y$: $\dfrac{\partial}{\partial y}\left(\dfrac{x}{x^2+y^2}\right)=x\times\dfrac{\partial}{\partial y}\left((x^2+y^2)^{-1}\right)=x\times\left(-(x^2+y^2)^{-2}\times2y\right)=\dfrac{-2xy}{(x^2+y^2)^2}$ (regola della catena, notebook 5.9, derivando prima la potenza $-1$ e poi l'interno $x^2+y^2$ rispetto a $y$).

**Passo 3 — valutiamo in $(1,1)$**: $\dfrac{\partial f}{\partial x}(1,1)=\dfrac{1-1}{(1+1)^2}=\dfrac04=0$. $\dfrac{\partial f}{\partial y}(1,1)=\dfrac{-2}{4}=-\dfrac12$.

**Conclusione:** in $(1,1)$, la superficie $z=f(x,y)$ è "piatta" nella direzione $x$ (derivata parziale nulla) ma scende nella direzione $y$ (derivata parziale negativa).

**Perché conviene sempre derivare tenendo l'altra variabile letteralmente "congelata" come un simbolo (non come un numero), anche quando alla fine valuteremo in un punto specifico:** derivare simbolicamente rispetto a $x$ con $y$ trattata come costante (ma non ancora sostituita con un valore numerico) produce una **funzione** $\partial f/\partial x(x,y)$, che possiamo poi valutare in **qualsiasi** punto senza dover rifare la derivazione da capo; se sostituissimo $y=1$ troppo presto (prima di derivare), perderemmo la possibilità di ottenere la formula generale — un errore metodologico comune quando si inizia a lavorare con più variabili.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = x / (x**2 + y**2)

df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
print("df/dx =", sp.simplify(df_dx))
print("df/dy =", sp.simplify(df_dy))

print("\ndf/dx(1,1) =", df_dx.subs({x: 1, y: 1}))
print("df/dy(1,1) =", df_dy.subs({x: 1, y: 1}))
````

## Riepilogo finale

- La **derivata parziale** $\partial f/\partial x$ si calcola derivando rispetto a $x$ trattando tutte le altre variabili come costanti — stesse regole del Modulo 5.
- Geometricamente, $\partial f/\partial x(x_0,y_0)$ è la pendenza della sezione del grafico di $f$ lungo il piano $y=y_0$.
- **Sorpresa importante**: l'esistenza di tutte le derivate parziali **non garantisce** la continuità — a differenza del caso unidimensionale (notebook 5.8), dove derivabilità implicava continuità.
- Questa "falla" delle derivate parziali motiva il concetto più robusto di differenziabilità, che studieremo nel prossimo notebook.
