# 7.13 — Moltiplicatori di Lagrange

**Prerequisiti:** 7.12 Massimi e minimi, 7.9 Gradiente, 6.2 Geometria vettoriale.


## Perché ci interessa

Nel notebook 7.12 abbiamo cercato massimi e minimi **liberi**, dove il punto poteva stare ovunque nel dominio. Ma spesso i problemi reali hanno **vincoli**: massimizzare il volume di una scatola con una quantità fissa di materiale per la superficie; minimizzare il costo di produzione dato un livello di output richiesto. In questi casi, il punto ottimo deve soddisfare un'equazione aggiuntiva $g(x,y)=c$ — restare su una curva o superficie, non muoversi liberamente in tutto il piano. Il metodo dei **moltiplicatori di Lagrange** risolve elegantemente questo problema, sfruttando proprio la proprietà geometrica del gradiente vista nel notebook 7.9.

## Teoria, passo dopo passo

### Il problema vincolato

Vogliamo trovare i punti di massimo o minimo di $f(x,y)$ **soggetti al vincolo** $g(x,y)=c$ (una curva di livello di $g$, notebook 7.4). Geometricamente: invece di muoverci liberamente sul dominio, siamo costretti a restare sulla curva $\{g=c\}$, e cerchiamo il punto più "alto" o più "basso" di $f$ **lungo quella curva**.

### L'intuizione geometrica: tangenza dei gradienti

Immaginiamo di muoverci lungo la curva $\{g=c\}$. Se, in un punto $x_0$ della curva, il vettore tangente alla curva **non** è perpendicolare a $\nabla f(x_0)$, allora muovendoci lungo la curva in una direzione o nell'altra $f$ cambia (aumenta o diminuisce) — quindi $x_0$ **non può** essere un massimo o minimo vincolato (potremmo sempre migliorare spostandoci lungo la curva). Un punto è candidato a essere un estremo vincolato solo se il vettore tangente alla curva **è** perpendicolare a $\nabla f(x_0)$.

Ma dal notebook 7.9 sappiamo che $\nabla g(x_0)$ è **sempre** perpendicolare alla curva $\{g=c\}$ (essendo essa stessa una curva di livello di $g$). Quindi: nel punto di ottimo, $\nabla f(x_0)$ deve essere perpendicolare alla stessa direzione tangente a cui $\nabla g(x_0)$ è perpendicolare — in $\mathbb R^2$, questo significa che $\nabla f(x_0)$ e $\nabla g(x_0)$ sono **paralleli** (entrambi perpendicolari alla stessa retta tangente, quindi allineati tra loro).

### Il teorema dei moltiplicatori di Lagrange

> **Teorema.** Se $f$ ha un estremo locale vincolato in $x_0$, soggetto a $g(x,y)=c$, e $\nabla g(x_0)\ne\mathbf 0$, allora esiste $\lambda\in\mathbb R$ (il **moltiplicatore di Lagrange**) tale che:
> $$\nabla f(x_0) = \lambda\,\nabla g(x_0)$$

Questa condizione, insieme al vincolo $g(x_0)=c$, dà un sistema di equazioni (in $\mathbb R^2$: 3 equazioni — le due componenti del gradiente più il vincolo — nelle 3 incognite $x,y,\lambda$) che permette di trovare i **candidati** a estremo vincolato.

### Procedimento pratico

1. Scrivi il sistema $\nabla f(x,y)=\lambda\nabla g(x,y)$, cioè $\partial f/\partial x=\lambda\,\partial g/\partial x$ e $\partial f/\partial y=\lambda\,\partial g/\partial y$.
2. Aggiungi il vincolo $g(x,y)=c$.
3. Risolvi il sistema (tipicamente eliminando $\lambda$ dalle prime due equazioni, poi sostituendo nel vincolo).
4. Valuta $f$ in ciascuna soluzione trovata: se il vincolo definisce un insieme **compatto** (notebook 7.3, 7.6), il teorema di Weierstrass garantisce che il massimo e il minimo assoluti vincolati esistono e si trovano tra questi candidati — basta confrontare i valori di $f$.

### Esempio guidato

Massimizza $f(x,y)=xy$ soggetto a $x^2+y^2=1$ (la circonferenza unitaria, un compatto). Poniamo $g(x,y)=x^2+y^2$, $c=1$. $\nabla f=(y,x)$, $\nabla g=(2x,2y)$. Il sistema: $y=2\lambda x$, $x=2\lambda y$, $x^2+y^2=1$. Dalle prime due: $y=2\lambda x$ e sostituendo nella seconda, $x=2\lambda(2\lambda x)=4\lambda^2x$, cioè $x(1-4\lambda^2)=0$. Se $x\ne0$: $\lambda^2=\frac14$, cioè $\lambda=\pm\frac12$; con $\lambda=\frac12$: $y=x$, e dal vincolo $2x^2=1$, $x=\pm\frac1{\sqrt2}$. Con $\lambda=-\frac12$: $y=-x$, $x=\pm\frac1{\sqrt2}$. Valutando $f=xy$ nei quattro candidati $\left(\pm\frac1{\sqrt2},\pm\frac1{\sqrt2}\right)$: massimo $\frac12$ (dove $x=y$), minimo $-\frac12$ (dove $x=-y$).

````python
import sympy as sp

x, y, lam = sp.symbols('x y lambda', real=True)

f = x * y
g = x**2 + y**2
c = 1

eq1 = sp.diff(f, x) - lam * sp.diff(g, x)
eq2 = sp.diff(f, y) - lam * sp.diff(g, y)
eq3 = g - c

soluzioni = sp.solve([eq1, eq2, eq3], [x, y, lam])
print("Candidati (x, y, lambda):")
for sol in soluzioni:
    print(" ", sol, " -> f =", f.subs({x: sol[0], y: sol[1]}))
````

## Esercizio svolto e commentato

**Problema:** tra tutti i rettangoli con perimetro fisso $P=20$, trova quello di area massima (un classico problema di ottimizzazione vincolata).

**Ragionamento:**

**Passo 1 — formalizziamo il problema.** Siano $x,y>0$ i lati del rettangolo. L'area è $f(x,y)=xy$ (da massimizzare); il vincolo è sul perimetro: $g(x,y)=2x+2y=20$, cioè $x+y=10$ (semplificando).

**Passo 2 — impostiamo il sistema di Lagrange**: $\nabla f=(y,x)$, $\nabla g=(2,2)$ (con $g(x,y)=2x+2y$). Il sistema: $y=2\lambda$, $x=2\lambda$, $2x+2y=20$.

**Passo 3 — dalle prime due equazioni**: $y=2\lambda=x$, cioè $x=y$ (il rettangolo ottimale è un **quadrato**, un risultato classico).

**Passo 4 — sostituiamo nel vincolo**: $2x+2x=20$, cioè $4x=20$, $x=5$; quindi $y=5$.

**Conclusione:** il rettangolo di area massima con perimetro $20$ è il quadrato $5\times5$, con area $f(5,5)=25$.

**Verifica che sia davvero un massimo (non un minimo o un punto sella vincolato):** il vincolo $x+y=10$ con $x,y>0$ è un segmento (non un insieme chiuso, essendo gli estremi $x=0$ o $y=0$ esclusi — corrisponderebbero a un "rettangolo" degenere, con area $0$); avvicinandosi agli estremi del segmento, l'area $f(x,y)=xy\to0$, mentre nel punto trovato $f=25>0$: per confronto diretto, $(5,5)$ dà il valore più alto tra i candidati sensati, confermando che è un massimo.

**Perché questo risultato — che tra tutti i rettangoli di perimetro fissato il quadrato ha area massima — è un fatto notevole (non solo un esercizio isolato):** è un caso particolare del principio più generale secondo cui, tra le figure con un vincolo di "misura totale" fissata (perimetro, superficie), le configurazioni **simmetriche** tendono a ottimizzare la quantità complementare (area, volume) — lo stesso principio, applicato ai solidi, spiega perché una sfera minimizza la superficie a parità di volume, un fatto che tornerà utile intuitivamente studiando problemi di ottimizzazione in fisica e ingegneria.

````python
import sympy as sp

x, y, lam = sp.symbols('x y lambda', positive=True)

f = x * y
g = 2*x + 2*y
c = 20

eq1 = sp.diff(f, x) - lam * sp.diff(g, x)
eq2 = sp.diff(f, y) - lam * sp.diff(g, y)
eq3 = g - c

soluzione = sp.solve([eq1, eq2, eq3], [x, y, lam])
print("Soluzione:", soluzione)
print("Area massima:", f.subs({x: soluzione[x], y: soluzione[y]}))
````

**4.** Se $\nabla g(x_0)=\mathbf 0$, il punto $x_0$ è un punto **critico** di $g$ stesso, non necessariamente un punto "regolare" della curva $\{g=c\}$ — la curva potrebbe avere una cuspide, un'auto-intersezione, o addirittura degenerare in un punto isolato lì. La dimostrazione del teorema (basata sul vettore tangente alla curva, ortogonale a $\nabla g$) presuppone che $\nabla g(x_0)\ne\mathbf 0$ per garantire che la curva abbia una tangente ben definita in quel punto; se questa ipotesi fallisce, il ragionamento geometrico crolla, e un estremo vincolato potrebbe verificarsi in quel punto **senza** che valga $\nabla f=\lambda\nabla g$ per nessun $\lambda$ — un caso speciale da controllare sempre separatamente, verificando dove $\nabla g$ si annulla lungo il vincolo.

## Riepilogo finale

- Per massimizzare/minimizzare $f$ soggetto a $g(x,y)=c$, i **moltiplicatori di Lagrange** richiedono $\nabla f(x_0)=\lambda\nabla g(x_0)$: il gradiente di $f$ deve essere parallelo al gradiente del vincolo.
- Geometricamente: nel punto di ottimo vincolato, la curva di livello di $f$ è **tangente** al vincolo (entrambi i gradienti perpendicolari alla stessa direzione tangente).
- Il metodo produce un sistema di equazioni (gradiente parallelo + vincolo) che dà **candidati**; confrontando $f$ nei candidati (su un vincolo compatto) si trova il vero massimo/minimo.
- L'ipotesi $\nabla g(x_0)\ne\mathbf 0$ è essenziale: nei punti dove il vincolo stesso è "singolare", il metodo può non individuare tutti gli estremi.
