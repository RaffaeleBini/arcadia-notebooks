# 7.2 — Norma e distanza

**Prerequisiti:** 7.1 Spazio euclideo, 6.2 Geometria vettoriale, 5.4 Limiti.


## Perché ci interessa

Per estendere il concetto di limite (notebook 5.4) a funzioni di più variabili, ci servirà dire con precisione cosa significa che un punto $x\in\mathbb R^n$ è "vicino" a un altro punto $x_0$. In una sola dimensione, la vicinanza si misurava con il valore assoluto $|x-x_0|$ (notebook 0.8). In $\mathbb R^n$ serve un analogo: la **norma** di un vettore e, a partire da essa, la **distanza** tra due punti. Questo notebook formalizza rigorosamente questi due concetti e dimostra le loro proprietà fondamentali — in particolare la disuguaglianza triangolare, cruciale per tutte le dimostrazioni epsilon-delta che faremo nei prossimi notebook.

## Teoria, passo dopo passo

### Norma euclidea

Per $x=(x_1,\dots,x_n)\in\mathbb R^n$, la **norma euclidea** è (già vista in $\mathbb R^2$ nel notebook 6.2, ora generalizzata):

$$\|x\| = \sqrt{x\cdot x} = \sqrt{x_1^2+x_2^2+\cdots+x_n^2}$$

(ben definita, per il notebook 7.1: $x\cdot x\ge0$ sempre, quindi la radice esiste).

### Disuguaglianza di Cauchy-Schwarz

> **Teorema (Cauchy-Schwarz).** Per ogni $x,y\in\mathbb R^n$: $|x\cdot y|\le\|x\|\,\|y\|$.

**Dimostrazione.** Se $y=\mathbf 0$, entrambi i membri sono $0$: la disuguaglianza vale banalmente. Supponiamo $y\ne\mathbf 0$. Per ogni $t\in\mathbb R$, consideriamo $p(t)=\|x-ty\|^2=(x-ty)\cdot(x-ty)$. Sviluppando con la bilinearità del prodotto scalare (notebook 6.20):

$$p(t) = x\cdot x - 2t(x\cdot y) + t^2(y\cdot y) = \|y\|^2t^2 - 2(x\cdot y)t + \|x\|^2$$

Questo è un polinomio di secondo grado in $t$ (notebook 1.6), con $\|y\|^2>0$ (essendo $y\ne\mathbf 0$). Poiché $p(t)=\|x-ty\|^2\ge0$ per **ogni** $t$ (una norma al quadrato non è mai negativa, notebook 7.1), il polinomio non può avere due radici reali distinte: il suo **discriminante** deve essere $\le0$ (altrimenti, tra le due radici, $p(t)$ sarebbe negativo). Il discriminante è $\Delta=4(x\cdot y)^2-4\|x\|^2\|y\|^2$; imponendo $\Delta\le0$:

$$4(x\cdot y)^2 \le 4\|x\|^2\|y\|^2 \quad\Longrightarrow\quad (x\cdot y)^2\le\|x\|^2\|y\|^2 \quad\Longrightarrow\quad |x\cdot y|\le\|x\|\|y\|$$

(nell'ultimo passaggio, estraendo la radice quadrata di entrambi i membri, entrambi non negativi). $\blacksquare$

### Proprietà della norma (incluso il triangolo)

Usando Cauchy-Schwarz, dimostriamo che la norma soddisfa tre proprietà fondamentali, per ogni $x,y\in\mathbb R^n$ e $\lambda\in\mathbb R$:

1. **positività**: $\|x\|\ge0$, con $\|x\|=0 \iff x=\mathbf 0$ (diretta conseguenza della positività del prodotto scalare, notebook 6.20, 7.1);
2. **omogeneità**: $\|\lambda x\|=|\lambda|\,\|x\|$ (segue da $\|\lambda x\|^2=(\lambda x)\cdot(\lambda x)=\lambda^2(x\cdot x)=\lambda^2\|x\|^2$, estraendo la radice);
3. **disuguaglianza triangolare**: $\|x+y\|\le\|x\|+\|y\|$.

**Dimostrazione della disuguaglianza triangolare.** $\|x+y\|^2=(x+y)\cdot(x+y)=\|x\|^2+2(x\cdot y)+\|y\|^2$. Per Cauchy-Schwarz, $x\cdot y\le|x\cdot y|\le\|x\|\|y\|$, quindi:

$$\|x+y\|^2 \le \|x\|^2+2\|x\|\|y\|+\|y\|^2 = (\|x\|+\|y\|)^2$$

Estraendo la radice quadrata (entrambi i membri non negativi, funzione crescente per $t\ge0$): $\|x+y\|\le\|x\|+\|y\|$. $\blacksquare$

### Distanza euclidea

La **distanza** tra due punti $x,y\in\mathbb R^n$ è $d(x,y)=\|x-y\|$ (già vista implicitamente in $\mathbb R^2$ nel notebook 2.3). Dalle proprietà della norma, discendono le proprietà attese di una distanza:

- $d(x,y)\ge0$, con uguaglianza solo per $x=y$;
- $d(x,y)=d(y,x)$ (simmetria: $\|x-y\|=\|-(y-x)\|=|-1|\|y-x\|=\|y-x\|$, per omogeneità);
- $d(x,z)\le d(x,y)+d(y,z)$ (disuguaglianza triangolare, applicando quella della norma a $x-z=(x-y)+(y-z)$).

Questa terza proprietà è quella che useremo costantemente: dice che "andare direttamente da $x$ a $z$" non è mai più lungo di "passare per un punto intermedio $y$".

````python
import numpy as np

x = np.array([3, -1, 2])
y = np.array([1, 4, -2])

norma_x = np.linalg.norm(x)
print("||x|| =", norma_x)

# Cauchy-Schwarz: |x.y| <= ||x|| ||y||
prodotto_scalare = np.dot(x, y)
print("\n|x.y| =", abs(prodotto_scalare))
print("||x||*||y|| =", np.linalg.norm(x) * np.linalg.norm(y))
print("Cauchy-Schwarz verificata?", abs(prodotto_scalare) <= np.linalg.norm(x) * np.linalg.norm(y))

# Disuguaglianza triangolare
somma = x + y
print("\n||x+y|| =", np.linalg.norm(somma))
print("||x||+||y|| =", np.linalg.norm(x) + np.linalg.norm(y))
print("Triangolare verificata?", np.linalg.norm(somma) <= np.linalg.norm(x) + np.linalg.norm(y))

# Distanza tra due punti e disuguaglianza triangolare per la distanza
z = np.array([0, 0, 0])
d_xz = np.linalg.norm(x - z)
d_xy = np.linalg.norm(x - y)
d_yz = np.linalg.norm(y - z)
print("\nd(x,z) =", d_xz, " | d(x,y)+d(y,z) =", d_xy + d_yz)
````

## Esercizio svolto e commentato

**Problema:** dati $x=(1,2,2)$ e $y=(4,0,3)$, verifica numericamente Cauchy-Schwarz e la disuguaglianza triangolare, poi trova un vettore $z\ne\mathbf 0$ per cui Cauchy-Schwarz vale con **uguaglianza**.

**Ragionamento:**

**Passo 1 — calcoliamo le norme**: $\|x\|=\sqrt{1+4+4}=\sqrt9=3$. $\|y\|=\sqrt{16+0+9}=\sqrt{25}=5$.

**Passo 2 — calcoliamo il prodotto scalare**: $x\cdot y=1\times4+2\times0+2\times3=4+0+6=10$.

**Passo 3 — verifichiamo Cauchy-Schwarz**: $|x\cdot y|=10$; $\|x\|\|y\|=3\times5=15$. Poiché $10\le15$, la disuguaglianza è verificata (e **stretta**, non è un'uguaglianza).

**Passo 4 — verifichiamo la disuguaglianza triangolare**: $x+y=(5,2,5)$, $\|x+y\|=\sqrt{25+4+25}=\sqrt{54}\approx7.35$. $\|x\|+\|y\|=8$. Poiché $7.35\le8$, verificata.

**Passo 5 — cerchiamo un vettore $z$ per cui vale l'uguaglianza in Cauchy-Schwarz.** Ripercorrendo la dimostrazione: l'uguaglianza $|x\cdot y|=\|x\|\|y\|$ si ha esattamente quando il discriminante di $p(t)$ è **zero**, cioè quando $p(t)=\|x-ty\|^2=0$ per qualche $t$: cioè quando $x=ty$ per qualche $t$, cioè quando $x,y$ sono **proporzionali** (paralleli — coerentemente con l'interpretazione geometrica del notebook 6.2, dove $\cos\theta=\pm1$ per vettori paralleli). Prendiamo $z=2x=(2,4,4)$: verifichiamo che $|x\cdot z|=\|x\|\|z\|$.

**Conclusione:** con $z=2x$, $x\cdot z=2(x\cdot x)=2\times9=18$; $\|x\|\|z\|=3\times\|2x\|=3\times6=18$: uguaglianza esatta.

**Perché la condizione di uguaglianza in Cauchy-Schwarz (vettori proporzionali) sarà utile più avanti nel corso:** questa condizione limite compare spesso in ottimizzazione — per esempio, il metodo dei moltiplicatori di Lagrange (notebook 7.13) userà proprio il fatto che due gradienti sono proporzionali nei punti di massimo/minimo vincolato, un'idea concettualmente identica a questa condizione di uguaglianza.

````python
import sympy as sp

x = sp.Matrix([1, 2, 2])
y = sp.Matrix([4, 0, 3])

norma_x = sp.sqrt(x.dot(x))
norma_y = sp.sqrt(y.dot(y))
print("||x|| =", norma_x, " ||y|| =", norma_y)
print("x.y =", x.dot(y))
print("Cauchy-Schwarz:", abs(x.dot(y)), "<=", norma_x * norma_y)

z = 2 * x
print("\nx.z =", x.dot(z), " ||x||*||z|| =", norma_x * sp.sqrt(z.dot(z)))
````

## Riepilogo finale

- La **norma** $\|x\|=\sqrt{x\cdot x}$ generalizza il valore assoluto a $\mathbb R^n$; la **distanza** $d(x,y)=\|x-y\|$ generalizza $|x-y|$.
- La **disuguaglianza di Cauchy-Schwarz** $|x\cdot y|\le\|x\|\|y\|$ è il risultato chiave da cui discende la disuguaglianza triangolare; vale con uguaglianza esattamente quando i vettori sono proporzionali.
- La norma soddisfa positività, omogeneità e disuguaglianza triangolare; la distanza soddisfa le analoghe proprietà di una metrica.
- Questi strumenti — norma e distanza — sono esattamente ciò che serve per dire rigorosamente "il punto $x$ è vicino al punto $x_0$", la base per definire i limiti in più variabili.
