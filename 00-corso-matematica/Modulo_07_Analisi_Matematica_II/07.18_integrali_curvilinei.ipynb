# 7.18 — Integrali curvilinei

**Prerequisiti:** 7.17 Integrali tripli, 6.1 Vettori, 5.8 Derivata, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Finora abbiamo integrato su regioni "piene" (intervalli, aree, volumi). Ma molte grandezze fisiche si accumulano lungo un **percorso**: il lavoro compiuto da una forza mentre un oggetto si muove lungo una traiettoria curva, la massa di un filo sottile piegato a forma di curva, la circolazione di un fluido attorno a un ostacolo. Per queste domande serve un nuovo strumento: l'**integrale curvilineo**, che integra lungo una curva invece che su una regione. Prima di definirlo, dobbiamo imparare a descrivere matematicamente una curva: la **parametrizzazione**.

## Teoria, passo dopo passo

### Curve parametriche

Una **curva parametrica** in $\mathbb R^n$ è una funzione vettoriale $\mathbf r:[a,b]\to\mathbb R^n$, $\mathbf r(t)=(x(t),y(t),\dots)$, dove ogni componente è una funzione di una variabile (notebook 3.1). Per esempio, $\mathbf r(t)=(\cos t,\sin t)$ per $t\in[0,2\pi]$ percorre la circonferenza unitaria (usando l'identità fondamentale del notebook 2.10: $\cos^2t+\sin^2t=1$ per ogni $t$, quindi ogni punto $\mathbf r(t)$ soddisfa $x^2+y^2=1$).

### Il vettore velocità

Se ogni componente di $\mathbf r(t)$ è derivabile (notebook 5.8), il **vettore velocità** (o tangente) è $\mathbf r'(t)=(x'(t),y'(t),\dots)$, calcolato derivando **componente per componente** — esattamente come si sommano o scalano i vettori (notebook 6.1). Geometricamente, $\mathbf r'(t)$ è tangente alla curva nel punto $\mathbf r(t)$, e la sua norma $\|\mathbf r'(t)\|$ (notebook 7.2) è la **velocità scalare** (quanto rapidamente si percorre la curva rispetto al parametro $t$).

### Lunghezza di una curva

La lunghezza di $\mathbf r(t)$ per $t\in[a,b]$ è (generalizzando l'idea di "somma di spostamenti infinitesimi", coerente col Modulo 5):

$$L = \int_a^b\|\mathbf r'(t)\|\,dt$$

**Esempio**: per la circonferenza $\mathbf r(t)=(\cos t,\sin t)$, $t\in[0,2\pi]$: $\mathbf r'(t)=(-\sin t,\cos t)$, $\|\mathbf r'(t)\|=\sqrt{\sin^2t+\cos^2t}=1$ (identità fondamentale, notebook 2.10). Quindi $L=\int_0^{2\pi}1\,dt=2\pi$: la circonferenza unitaria!

### Integrale curvilineo di una funzione scalare

Dato $f(x,y)$ e una curva $\mathbf r(t)$, $t\in[a,b]$, l'**integrale curvilineo (di prima specie)** è:

$$\int_Cf\,ds = \int_a^bf(\mathbf r(t))\,\|\mathbf r'(t)\|\,dt$$

Il fattore $\|\mathbf r'(t)\|dt$ (spesso scritto $ds$, l'elemento di lunghezza d'arco) è il "prezzo" per riparametrizzare l'integrazione dalla lunghezza fisica al parametro $t$ — esattamente lo stesso ruolo del fattore Jacobiano nel notebook 7.15, qui in una versione unidimensionale. Con $f=1$, si ritrova la formula della lunghezza.

### Integrale curvilineo di un campo vettoriale (lavoro)

Dato un campo vettoriale $\mathbf F(x,y)=(P(x,y),Q(x,y))$ (che formalizzeremo nel notebook 7.20; per ora, pensalo come una forza che dipende dalla posizione), il **lavoro** compiuto da $\mathbf F$ lungo la curva $\mathbf r(t)$ è:

$$\int_C\mathbf F\cdot d\mathbf r = \int_a^b\mathbf F(\mathbf r(t))\cdot\mathbf r'(t)\,dt$$

(prodotto scalare, notebook 6.2, tra la forza e il vettore velocità: il lavoro è "forza per spostamento nella direzione del moto", generalizzato a un percorso curvo tramite l'integrazione). Notazione alternativa: $\int_CP\,dx+Q\,dy$, esplicitando le componenti.

````python
import sympy as sp

t = sp.symbols('t', real=True)

# Curva: circonferenza unitaria
r = sp.Matrix([sp.cos(t), sp.sin(t)])
r_prime = r.diff(t)
print("r(t) =", r.T, " | r'(t) =", r_prime.T)

norma_r_prime = sp.sqrt((r_prime.T * r_prime)[0])
norma_r_prime = sp.simplify(norma_r_prime)
print("||r'(t)|| =", norma_r_prime)

lunghezza = sp.integrate(norma_r_prime, (t, 0, 2*sp.pi))
print("Lunghezza della circonferenza:", lunghezza)

# Integrale curvilineo di un campo vettoriale: F(x,y)=(y,-x) lungo la circonferenza
x, y = sp.symbols('x y', real=True)
F = sp.Matrix([y, -x])
F_lungo_curva = F.subs({x: r[0], y: r[1]})
integranda = (F_lungo_curva.T * r_prime)[0]
print("\nF lungo la curva:", F_lungo_curva.T)
print("Integranda F.r'(t):", sp.simplify(integranda))

lavoro = sp.integrate(sp.simplify(integranda), (t, 0, 2*sp.pi))
print("Lavoro (integrale curvilineo):", lavoro)
````

## Esercizio svolto e commentato

**Problema:** calcola il lavoro compiuto dal campo $\mathbf F(x,y)=(x,y)$ lungo la curva $\mathbf r(t)=(t,t^2)$, $t\in[0,1]$ (un arco di parabola).

**Ragionamento:**

**Passo 1 — calcoliamo il vettore velocità**: $\mathbf r'(t)=(1,2t)$ (derivando ogni componente).

**Passo 2 — valutiamo $\mathbf F$ lungo la curva**, sostituendo $x=t,\,y=t^2$: $\mathbf F(\mathbf r(t))=(t,t^2)$.

**Passo 3 — calcoliamo il prodotto scalare** $\mathbf F(\mathbf r(t))\cdot\mathbf r'(t)$: $(t,t^2)\cdot(1,2t)=t\times1+t^2\times2t=t+2t^3$.

**Passo 4 — integriamo rispetto a $t$**: $\int_0^1(t+2t^3)dt=\left[\dfrac{t^2}2+\dfrac{t^4}2\right]_0^1=\dfrac12+\dfrac12=1$.

**Conclusione:** il lavoro è $1$.

**Perché, per questo campo specifico $\mathbf F(x,y)=(x,y)$, il risultato dipende solo dai punti estremi della curva e non dalla forma del percorso (verificabile confrontando col calcolo, nella cella di codice sotto, dello stesso campo lungo il segmento rettilineo $\mathbf r(t)=(t,t)$, $t\in[0,1]$ — che va dallo stesso punto iniziale $(0,0)$ allo stesso punto finale $(1,1)$ della parabola):** questo campo è il **gradiente** di $g(x,y)=\frac12(x^2+y^2)$ (verifica: $\nabla g=(x,y)=\mathbf F$); per campi che sono gradienti (detti **conservativi**), esiste un risultato notevole — il lavoro dipende solo dai punti estremi, non dal percorso — che formalizzeremo con il teorema fondamentale per gli integrali curvilinei nel notebook 7.20. Questo spiega perché il valore $1$ coincide esattamente con $g(1,1)-g(0,0)=\frac12(1+1)-0=1$: una coincidenza che, in realtà, è tutt'altro che casuale.

````python
import sympy as sp

t = sp.symbols('t', real=True)

r = sp.Matrix([t, t**2])
r_prime = r.diff(t)

x, y = sp.symbols('x y', real=True)
F = sp.Matrix([x, y])
F_lungo_curva = F.subs({x: r[0], y: r[1]})

integranda = (F_lungo_curva.T * r_prime)[0]
print("Integranda:", integranda)

lavoro = sp.integrate(integranda, (t, 0, 1))
print("Lavoro lungo la parabola:", lavoro)

# Confronto: stesso campo, lungo il segmento rettilineo (0,0)->(1,1)
r_retta = sp.Matrix([t, t])
r_retta_prime = r_retta.diff(t)
F_lungo_retta = F.subs({x: r_retta[0], y: r_retta[1]})
integranda_retta = (F_lungo_retta.T * r_retta_prime)[0]
lavoro_retta = sp.integrate(integranda_retta, (t, 0, 1))
print("Lavoro lungo il segmento (stesso campo, stessi estremi):", lavoro_retta)
````

## Riepilogo finale

- Una **curva parametrica** $\mathbf r(t)$ ha vettore velocità $\mathbf r'(t)$ (derivata componente per componente); la sua norma è la velocità scalare, e $\int_a^b\|\mathbf r'(t)\|dt$ dà la lunghezza della curva.
- L'**integrale curvilineo di prima specie** $\int_Cf\,ds=\int_a^bf(\mathbf r(t))\|\mathbf r'(t)\|dt$ integra una funzione scalare lungo una curva, pesata per la lunghezza d'arco.
- L'**integrale curvilineo di un campo vettoriale** $\int_C\mathbf F\cdot d\mathbf r=\int_a^b\mathbf F(\mathbf r(t))\cdot\mathbf r'(t)\,dt$ calcola il lavoro compiuto da $\mathbf F$ lungo la curva.
- Per campi **conservativi** (gradienti di una funzione), il lavoro dipende solo dai punti estremi, non dal percorso — un'anteprima del teorema fondamentale per gli integrali curvilinei.
