# 7.3 — Topologia elementare di $\mathbb R^n$

**Prerequisiti:** 7.2 Norma e distanza, 0.14 Intervalli, 5.2 Completezza dei reali.


## Perché ci interessa

Nel notebook 0.14 abbiamo classificato gli intervalli di $\mathbb R$ come aperti, chiusi, o né l'uno né l'altro — una distinzione che si è rivelata cruciale nel Modulo 5 per teoremi come Weierstrass (che richiede un intervallo chiuso e limitato). Per estendere quei teoremi a più variabili, dobbiamo prima generalizzare le nozioni di "aperto" e "chiuso" da intervalli di $\mathbb R$ a sottoinsiemi di $\mathbb R^n$. Lo strumento chiave è la **palla aperta**, l'analogo multidimensionale dell'intervallo aperto — costruita proprio con la distanza definita nel notebook precedente.

## Teoria, passo dopo passo

### La palla aperta

Dato un punto $x_0\in\mathbb R^n$ e un raggio $r>0$, la **palla aperta** di centro $x_0$ e raggio $r$ è:

$$B(x_0,r) = \{x\in\mathbb R^n : d(x,x_0)<r\} = \{x\in\mathbb R^n : \|x-x_0\|<r\}$$

Per $n=1$, $B(x_0,r)=(x_0-r,x_0+r)$: esattamente un intervallo aperto (notebook 0.14)! Per $n=2$, è un disco aperto (senza il bordo); per $n=3$, una sfera piena aperta (senza la superficie).

### Insiemi aperti

Un insieme $A\subseteq\mathbb R^n$ è **aperto** se ogni suo punto è "circondato" interamente da $A$: per ogni $x\in A$, esiste $r>0$ tale che $B(x,r)\subseteq A$. Intuitivamente: partendo da un punto di $A$, ci si può sempre muovere un po' in **qualsiasi** direzione restando dentro $A$ — nessun punto di $A$ è "sul bordo".

**Esempio**: $A=\{(x,y):x^2+y^2<1\}$ (il disco unitario aperto) è aperto. Presa una circonferenza di raggio $r=1-\|x_0\|$ (positivo, perché $\|x_0\|<1$) attorno a un punto $x_0\in A$ qualsiasi, per la disuguaglianza triangolare (notebook 7.2) ogni punto della palla $B(x_0,r)$ dista dall'origine meno di $\|x_0\|+r=1$: resta dentro $A$.

### Insiemi chiusi

Un insieme $C\subseteq\mathbb R^n$ è **chiuso** se il suo complementare $\mathbb R^n\setminus C$ è aperto. Equivalentemente (e più intuitivamente): $C$ è chiuso se contiene tutti i suoi **punti di frontiera** (o, come vedremo nel Modulo 10, se contiene tutti i limiti di successioni di suoi elementi).

**Esempio**: $\overline C=\{(x,y):x^2+y^2\le1\}$ (il disco unitario **chiuso**, con il bordo) è chiuso: il suo complementare $\{x^2+y^2>1\}$ è aperto (verifica analoga a sopra).

### Attenzione: aperto e chiuso non sono opposti

Come per gli intervalli (notebook 0.14), un insieme può essere **né** aperto **né** chiuso (esempio: $\{(x,y):0<x\le1,\,0\le y<1\}$, aperto in una direzione e chiuso nell'altra), oppure — caso limite — **sia** aperto **sia** chiuso: $\emptyset$ e $\mathbb R^n$ stesso sono gli unici due esempi di questo tipo.

### Interno, frontiera e chiusura

Dato $A\subseteq\mathbb R^n$: l'**interno** $\operatorname{int}(A)$ è l'insieme dei punti $x\in A$ per cui esiste una palla $B(x,r)\subseteq A$ (i punti "sicuramente dentro"); la **frontiera** $\partial A$ è l'insieme dei punti $x$ tali che **ogni** palla $B(x,r)$ contiene sia punti di $A$ sia punti del complementare (i punti "di confine", che possono appartenere o no ad $A$); la **chiusura** $\overline A=A\cup\partial A$. $A$ è aperto se e solo se $A=\operatorname{int}(A)$; $A$ è chiuso se e solo se $A=\overline A$.

### Insiemi limitati e compatti

$A$ è **limitato** se esiste $M>0$ tale che $\|x\|\le M$ per ogni $x\in A$ (sta dentro una palla abbastanza grande) — generalizzazione diretta del concetto già visto per successioni e funzioni nel Modulo 4-5. Un insieme **chiuso e limitato** in $\mathbb R^n$ si dice **compatto**: questi insiemi giocheranno lo stesso ruolo cruciale che gli intervalli chiusi e limitati $[a,b]$ giocavano nel teorema di Weierstrass (notebook 5.7) — il prossimo notebook (7.6) enuncerà la versione multidimensionale di quel teorema, valida esattamente sui compatti.

````python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

theta = np.linspace(0, 2*np.pi, 200)
cerchio_x, cerchio_y = np.cos(theta), np.sin(theta)

# Disco aperto: bordo tratteggiato (escluso)
axes[0].fill(cerchio_x, cerchio_y, alpha=0.3, color='blue')
axes[0].plot(cerchio_x, cerchio_y, '--', color='blue')
axes[0].set_title("Disco APERTO: {x^2+y^2 < 1}\n(bordo escluso, tratteggiato)")
axes[0].set_aspect('equal')

# Disco chiuso: bordo continuo (incluso)
axes[1].fill(cerchio_x, cerchio_y, alpha=0.3, color='red')
axes[1].plot(cerchio_x, cerchio_y, '-', color='red')
axes[1].set_title("Disco CHIUSO: {x^2+y^2 <= 1}\n(bordo incluso, continuo)")
axes[1].set_aspect('equal')

plt.tight_layout()
plt.show()

# Verifica numerica: un punto vicino al bordo del disco aperto ha ancora una palla dentro A
import numpy as np
x0 = np.array([0.9, 0])  # dentro il disco aperto (norma 0.9 < 1)
raggio_sicuro = 1 - np.linalg.norm(x0)
print("Punto x0 =", x0, " norma =", np.linalg.norm(x0))
print("Raggio della palla garantita dentro il disco aperto:", raggio_sicuro)
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $A=\{(x,y)\in\mathbb R^2 : 1<x^2+y^2\le4\}$ (una "corona circolare") è aperto, chiuso, né l'uno né l'altro; trova la sua frontiera.

**Ragionamento:**

**Passo 1 — controlliamo se $A$ è aperto**, verificando un punto sul bordo interno: prendiamo $x_0$ con $\|x_0\|=1$... ma attenzione, $\|x_0\|=1$ **non** soddisfa $1<x^2+y^2$ (la disuguaglianza è stretta): quel punto non appartiene ad $A$. Prendiamo invece un punto **appena dentro**, con $\|x_0\|=2$ (sul bordo esterno, che invece **è incluso**, per la disuguaglianza $\le4$): per **ogni** $r>0$, la palla $B(x_0,r)$ contiene punti con $\|x\|>2$, che non soddisfano $x^2+y^2\le4$, quindi non appartengono ad $A$. Nessuna palla centrata in $x_0$ è interamente contenuta in $A$: $A$ **non è aperto**.

**Passo 2 — controlliamo se $A$ è chiuso**, verificando il complementare vicino al bordo interno: prendiamo $x_0$ con $\|x_0\|=1$ (che, come notato, **non** appartiene ad $A$, quindi appartiene al complementare). Per ogni $r>0$, la palla $B(x_0,r)$ contiene punti con $\|x\|$ leggermente **maggiore** di $1$ (quindi con $x^2+y^2>1$) che, se anche $\le4$, **appartengono** ad $A$: nessuna palla attorno a $x_0$ è contenuta interamente nel complementare. Il complementare non è aperto, quindi $A$ **non è chiuso**.

**Conclusione:** $A$ non è né aperto né chiuso — ha un pezzo di frontiera **escluso** (la circonferenza interna, $\|x\|=1$) e un pezzo **incluso** (la circonferenza esterna, $\|x\|=2$). La frontiera completa è $\partial A=\{\|x\|=1\}\cup\{\|x\|=2\}$ (entrambe le circonferenze, indipendentemente dal fatto che appartengano o no ad $A$: la frontiera è definita in base a **entrambi** i lati, non in base all'appartenenza).

**Perché questo esempio, con un bordo incluso e uno escluso, è istruttivo:** mostra che, esattamente come per gli intervalli del tipo $(a,b]$ nel notebook 0.14 (né aperti né chiusi), la topologia di $\mathbb R^n$ ammette insiemi "misti"; la definizione formale (basata su palle, non su intuizione visiva) permette di classificarli con precisione anche quando la figura potrebbe sembrare ambigua a colpo d'occhio.

````python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 300)
r_esterno = 2 * np.column_stack([np.cos(theta), np.sin(theta)])
r_interno = 1 * np.column_stack([np.cos(theta), np.sin(theta)])

fig, ax = plt.subplots(figsize=(5, 5))
ax.fill(np.concatenate([r_esterno[:,0], r_interno[::-1,0]]),
        np.concatenate([r_esterno[:,1], r_interno[::-1,1]]),
        alpha=0.3, color='purple')
ax.plot(r_esterno[:,0], r_esterno[:,1], '-', color='purple', label='bordo esterno (incluso)')
ax.plot(r_interno[:,0], r_interno[:,1], '--', color='purple', label='bordo interno (escluso)')
ax.set_aspect('equal')
ax.legend()
ax.set_title("A = {1 < x^2+y^2 <= 4}: né aperto né chiuso")
plt.show()
````

## Esercizi da risolvere

1. $A=\{(x,y):x>0\}$ (semipiano destro, senza l'asse $y$): è aperto? È chiuso?
2. $A=\{(x,y):x\ge0\}$ (semipiano destro, con l'asse $y$ incluso): è aperto? È chiuso?
3. Trova la frontiera $\partial A$ per $A=\{(x,y):x^2+y^2<1\}\cup\{(2,0)\}$ (il disco aperto, unito a un punto isolato).
4. *(più impegnativo)* Dimostra che l'intersezione di due insiemi aperti $A,B\subseteq\mathbb R^n$ è aperta (suggerimento: per $x\in A\cap B$, trova due raggi $r_A,r_B$ e usa il minore dei due).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Per ogni punto con $x>0$, si può trovare una palla piccola abbastanza (raggio $=x$) interamente contenuta nel semipiano: **aperto**. Il complementare $\{x\le0\}$ non è aperto (per gli stessi motivi, applicati a un punto con $x=0$: ogni palla attorno contiene punti con $x>0$), quindi $A$ **non è chiuso**.

**2.** Il punto $(0,0)\in A$: ogni palla attorno contiene punti con $x<0$, non in $A$: **non aperto**. Il complementare $\{x<0\}$ è aperto (stesso ragionamento del punto 1): quindi $A$ **è chiuso**.

**3.** $\partial A$ include la circonferenza $\{x^2+y^2=1\}$ (bordo del disco: ogni palla attorno a un suo punto contiene sia punti dentro sia fuori dal disco) **e** il punto isolato $(2,0)$ stesso (ogni palla attorno a $(2,0)$, per raggio sufficientemente piccolo, contiene sia $(2,0)$ — che appartiene ad $A$ — sia punti che non vi appartengono, essendo $(2,0)$ isolato dal resto di $A$).

**4.** Sia $x\in A\cap B$. Poiché $x\in A$ (aperto), esiste $r_A>0$ con $B(x,r_A)\subseteq A$. Poiché $x\in B$ (aperto), esiste $r_B>0$ con $B(x,r_B)\subseteq B$. Prendendo $r=\min(r_A,r_B)>0$: $B(x,r)\subseteq B(x,r_A)\subseteq A$ e $B(x,r)\subseteq B(x,r_B)\subseteq B$ (una palla di raggio più piccolo è sempre contenuta in una di raggio più grande, stesso centro), quindi $B(x,r)\subseteq A\cap B$. Questo vale per ogni $x\in A\cap B$: $A\cap B$ è aperto.

## Riepilogo finale

- La **palla aperta** $B(x_0,r)=\{x:\|x-x_0\|<r\}$ generalizza l'intervallo aperto a $\mathbb R^n$.
- $A$ è **aperto** se ogni suo punto ha una palla interamente contenuta in $A$; $A$ è **chiuso** se il suo complementare è aperto.
- Un insieme può essere né aperto né chiuso (come per gli intervalli, notebook 0.14), o entrambi (solo $\emptyset$ e $\mathbb R^n$).
- Un insieme **compatto** (chiuso e limitato) è l'analogo multidimensionale dell'intervallo $[a,b]$, e sarà l'ipotesi chiave per la versione multidimensionale del teorema di Weierstrass.
