# 7.21 — Divergenza

**Prerequisiti:** 7.20 Campi vettoriali, 7.7 Derivate parziali.


## Perché ci interessa

Immaginiamo un campo vettoriale come il flusso di un fluido: in ogni punto, $\mathbf F(x,y,z)$ indica direzione e velocità del fluido lì. Una domanda naturale: in un dato punto, il fluido si sta **espandendo** (come vicino a una sorgente) o **comprimendo** (come vicino a uno scarico)? La **divergenza** risponde esattamente a questa domanda: un numero (non un vettore) che misura il tasso netto di "espansione" del campo in un punto. È il primo dei due operatori differenziali fondamentali sui campi vettoriali (l'altro, il rotore, arriva nel prossimo notebook), e sarà l'ingrediente chiave del teorema della divergenza (notebook 7.24).

## Teoria, passo dopo passo

### Definizione

Per un campo vettoriale $\mathbf F(x,y,z)=(P,Q,R)$ con componenti derivabili, la **divergenza** è la funzione scalare:

$$\operatorname{div}\mathbf F = \frac{\partial P}{\partial x}+\frac{\partial Q}{\partial y}+\frac{\partial R}{\partial z}$$

(in $\mathbb R^2$, analogamente, $\operatorname{div}\mathbf F=\partial P/\partial x+\partial Q/\partial y$ per $\mathbf F=(P,Q)$). Si usa spesso la notazione $\nabla\cdot\mathbf F$ (un "prodotto scalare formale" tra l'operatore $\nabla=(\partial/\partial x,\partial/\partial y,\partial/\partial z)$ e il campo, notebook 6.2 — un'analogia utile mnemonica, non un vero prodotto scalare tra vettori numerici).

### Interpretazione fisica: tasso di espansione

Consideriamo un campo di velocità di un fluido. Se $\operatorname{div}\mathbf F(x_0)>0$, il punto $x_0$ è vicino a una "sorgente": il fluido si allontana da lì più di quanto vi arrivi (il flusso netto uscente da un piccolo volume attorno a $x_0$ è positivo). Se $\operatorname{div}\mathbf F(x_0)<0$, è vicino a uno "scarico" (pozzo): il flusso netto è entrante. Se $\operatorname{div}\mathbf F=0$ ovunque, il campo si dice **solenoidale** (o a divergenza nulla): il fluido non si crea né si distrugge, si limita a "circolare" (come un fluido incomprimibile).

### Esempio: campo radiale

$\mathbf F(x,y)=(x,y)$ (già incontrato nei notebook 7.18, 7.20 come campo conservativo): $\operatorname{div}\mathbf F=\partial x/\partial x+\partial y/\partial y=1+1=2>0$ — ovunque positiva, coerente con l'interpretazione di questo campo come "espansione radiale" verso l'esterno (ogni punto si allontana dall'origine, notebook 7.9, esercizio 4).

### Esempio: campo rotazionale (a divergenza nulla)

$\mathbf F(x,y)=(-y,x)$ (visto nel notebook 7.18, non conservativo): $\operatorname{div}\mathbf F=\partial(-y)/\partial x+\partial x/\partial y=0+0=0$. Questo campo, che "ruota" attorno all'origine (come le lancette di un orologio, in verso antiorario), ha divergenza nulla ovunque: il fluido "circola" senza espandersi né contrarsi, coerente con l'intuizione che una pura rotazione non crea né distrugge materia.

### Divergenza del gradiente: il Laplaciano

Un caso speciale importante: se $\mathbf F=\nabla f$ (campo conservativo, notebook 7.20), la divergenza del gradiente si chiama **Laplaciano** di $f$:

$$\Delta f := \operatorname{div}(\nabla f) = \frac{\partial^2f}{\partial x^2}+\frac{\partial^2f}{\partial y^2}+\frac{\partial^2f}{\partial z^2}$$

(la somma delle derivate parziali seconde "pure", notebook 7.10 — non quelle miste). Funzioni con $\Delta f=0$ (dette **armoniche**) descrivono, per esempio, la distribuzione di temperatura all'equilibrio in un corpo senza sorgenti di calore interne: un concetto che ricorrerà nel Modulo 15 (equazioni di Laplace e Poisson).

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)

def divergenza(F, variabili):
    return sum(sp.diff(F[i], variabili[i]) for i in range(len(F)))

# Campo radiale
F1 = [x, y]
print("div(x,y) =", divergenza(F1, [x, y]))

# Campo rotazionale
F2 = [-y, x]
print("div(-y,x) =", divergenza(F2, [x, y]))

# Laplaciano di f(x,y) = x^2+y^2 (paraboloide)
f = x**2 + y**2
grad_f = [sp.diff(f, x), sp.diff(f, y)]
laplaciano = divergenza(grad_f, [x, y])
print("\nLaplaciano di x^2+y^2:", laplaciano)

# Una funzione armonica: f(x,y) = x^2 - y^2
f_armonica = x**2 - y**2
grad_f2 = [sp.diff(f_armonica, x), sp.diff(f_armonica, y)]
print("Laplaciano di x^2-y^2:", divergenza(grad_f2, [x, y]), "-> armonica (Laplaciano nullo)")
````

## Esercizio svolto e commentato

**Problema:** calcola la divergenza di $\mathbf F(x,y,z)=(x^2y,\,e^{yz},\,\sin(xz))$, e valutala nel punto $(1,0,\pi)$.

**Ragionamento:**

**Passo 1 — identifichiamo le componenti**: $P=x^2y$, $Q=e^{yz}$, $R=\sin(xz)$.

**Passo 2 — calcoliamo $\partial P/\partial x$**, derivando rispetto a $x$ trattando $y$ come costante: $\partial P/\partial x=2xy$.

**Passo 3 — calcoliamo $\partial Q/\partial y$**, derivando rispetto a $y$ trattando $z$ come costante: $\partial Q/\partial y=ze^{yz}$ (regola della catena, notebook 5.9).

**Passo 4 — calcoliamo $\partial R/\partial z$**, derivando rispetto a $z$ trattando $x$ come costante: $\partial R/\partial z=x\cos(xz)$.

**Passo 5 — sommiamo** (definizione di divergenza): $\operatorname{div}\mathbf F=2xy+ze^{yz}+x\cos(xz)$.

**Passo 6 — valutiamo in $(1,0,\pi)$**: $2\times1\times0+\pi\times e^{0\times\pi}+1\times\cos(1\times\pi)=0+\pi\times1+\cos\pi=\pi+(-1)=\pi-1$.

**Conclusione:** $\operatorname{div}\mathbf F(1,0,\pi)=\pi-1\approx2.14>0$: in quel punto, il campo si comporta localmente come una "sorgente".

**Perché calcolare ciascuna derivata parziale isolatamente (passi 2-4), prima di sommarle (passo 5), riduce il rischio di errore rispetto a calcolare tutto insieme:** la divergenza richiede tre derivazioni **diverse**, ciascuna rispetto a una variabile diversa e su una componente diversa del campo — trattarle separatamente, come nei passi 2-4, permette di applicare con calma le regole di derivazione (notebook 7.7) a ciascun pezzo, evitando di confondere quale variabile trattare come costante in quale termine, un errore comune quando si combinano più derivazioni contemporaneamente in un unico passaggio.

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)
P = x**2 * y
Q = sp.exp(y * z)
R = sp.sin(x * z)

div_F = sp.diff(P, x) + sp.diff(Q, y) + sp.diff(R, z)
print("div(F) =", div_F)

valore = div_F.subs({x: 1, y: 0, z: sp.pi})
print("div(F)(1,0,pi) =", sp.simplify(valore))
````

**1.** $\partial(xyz)/\partial x=yz$. $\partial(x^2)/\partial y=0$. $\partial(y^2z)/\partial z=y^2$. $\operatorname{div}\mathbf F=yz+0+y^2=yz+y^2$.

**2.** $P=x^3-3xy^2$, $Q=3x^2y-y^3$. $\partial P/\partial x=3x^2-3y^2$. $\partial Q/\partial y=3x^2-3y^2$. $\operatorname{div}\mathbf F=3x^2-3y^2+3x^2-3y^2=6x^2-6y^2$: **non è identicamente nulla** (per esempio, in $(1,0)$ vale $6\ne0$): non è solenoidale.

**3.** $f_x=\dfrac{2x}{x^2+y^2}$, $f_{xx}=\dfrac{2(x^2+y^2)-2x\times2x}{(x^2+y^2)^2}=\dfrac{2y^2-2x^2}{(x^2+y^2)^2}$. Per simmetria (scambiando $x\leftrightarrow y$): $f_{yy}=\dfrac{2x^2-2y^2}{(x^2+y^2)^2}$. Somma: $\Delta f=f_{xx}+f_{yy}=\dfrac{2y^2-2x^2+2x^2-2y^2}{(x^2+y^2)^2}=0$: **armonica** (per $(x,y)\ne(0,0)$, dove la funzione non è definita).

**4.** Il rotore (anticipato) ha componenti $\left(\dfrac{\partial R}{\partial y}-\dfrac{\partial Q}{\partial z},\ \dfrac{\partial P}{\partial z}-\dfrac{\partial R}{\partial x},\ \dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}\right)$. La sua divergenza è la somma delle derivate di ciascuna componente rispetto alla propria variabile: $\dfrac{\partial}{\partial x}\left(\dfrac{\partial R}{\partial y}-\dfrac{\partial Q}{\partial z}\right)+\dfrac{\partial}{\partial y}\left(\dfrac{\partial P}{\partial z}-\dfrac{\partial R}{\partial x}\right)+\dfrac{\partial}{\partial z}\left(\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}\right)$. Espandendo, ogni derivata seconda mista (es. $\partial^2R/\partial x\partial y$) compare esattamente **due volte con segno opposto** (per il teorema di Schwarz, notebook 7.10, l'ordine di derivazione non conta, quindi i due termini sono numericamente uguali e si cancellano): la somma totale è $0$.

## Riepilogo finale

- La **divergenza** $\operatorname{div}\mathbf F=\partial P/\partial x+\partial Q/\partial y+\partial R/\partial z$ è una funzione **scalare** che misura il tasso di espansione/contrazione di un campo in un punto.
- Divergenza positiva $\to$ sorgente locale; negativa $\to$ pozzo locale; nulla ovunque $\to$ campo **solenoidale**.
- Il **Laplaciano** $\Delta f=\operatorname{div}(\nabla f)$ è la divergenza del gradiente di una funzione scalare; funzioni con $\Delta f=0$ si dicono **armoniche**.
- Un fatto generale (verificato nell'esercizio 4): la divergenza del rotore di un campo è sempre nulla — un'identità che riprenderemo nel Modulo 15 e nella teoria dei campi vettoriali.
