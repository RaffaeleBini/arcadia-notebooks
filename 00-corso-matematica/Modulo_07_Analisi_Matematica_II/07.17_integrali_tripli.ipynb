# 7.17 — Integrali tripli

**Prerequisiti:** 7.16 Integrali doppi, 7.15 Cambiamento di variabili.


## Perché ci interessa

Tutto quanto costruito per gli integrali doppi (Fubini, cambiamento di variabili, applicazioni fisiche) si estende naturalmente a tre variabili. L'**integrale triplo** $\iiint_Ef(x,y,z)\,dV$ su una regione solida $E\subset\mathbb R^3$ calcola grandezze come la massa di un solido con densità variabile, o semplicemente il suo volume — uno strumento essenziale in fisica (momenti d'inerzia, campi gravitazionali) e ingegneria.

## Teoria, passo dopo passo

### Definizione e teorema di Fubini (tre variabili)

Esattamente come nel notebook 7.14, l'integrale triplo si costruisce come limite di somme su una griglia tridimensionale di piccoli parallelepipedi. Il teorema di Fubini si estende: per $E=[a,b]\times[c,d]\times[p,q]$ (un parallelepipedo):

$$\iiint_Ef\,dV = \int_a^b\int_c^d\int_p^qf(x,y,z)\,dz\,dy\,dx$$

(in un ordine qualsiasi tra le tre variabili, il risultato è lo stesso). Per regioni solide più generali, si procede come nel notebook 7.14: si descrive $E$ con limiti di integrazione via via più "interni" dipendenti dalle variabili "più esterne" — per esempio $E=\{(x,y,z):(x,y)\in D,\ \psi_1(x,y)\le z\le\psi_2(x,y)\}$, integrando prima $z$ (tra due superfici), poi $(x,y)$ su una regione $D$ del piano (con le tecniche del notebook 7.16).

### Volume e massa

Ponendo $f=1$: $\iiint_E1\,dV$ dà il **volume** di $E$. Con una densità $\rho(x,y,z)$: $M=\iiint_E\rho\,dV$ dà la **massa**, e il baricentro si calcola con formule analoghe a quelle del notebook 7.16, in tre componenti $(\bar x,\bar y,\bar z)$.

### Coordinate cilindriche

Le **coordinate cilindriche** $(r,\theta,z)$ combinano le coordinate polari (nel piano $xy$) con la coordinata $z$ invariata: $x=r\cos\theta$, $y=r\sin\theta$, $z=z$. Utili per solidi con simmetria attorno a un asse (cilindri, coni). Lo Jacobiano si calcola come nel notebook 7.15, ma ora è una matrice $3\times3$:

$$J = \begin{pmatrix}\cos\theta&-r\sin\theta&0\\\sin\theta&r\cos\theta&0\\0&0&1\end{pmatrix} \qquad \det J = r$$

(sviluppando lungo l'ultima riga, notebook 6.12: si riduce al determinante $2\times2$ già calcolato per le coordinate polari, moltiplicato per $1$). Quindi $dV=r\,dr\,d\theta\,dz$.

### Coordinate sferiche

Le **coordinate sferiche** $(\rho,\varphi,\theta)$ (con $\rho$ la distanza dall'origine, $\varphi$ l'angolo dal semiasse $z$ positivo, $\theta$ l'angolo azimutale come nelle polari) sono: $x=\rho\sin\varphi\cos\theta$, $y=\rho\sin\varphi\sin\theta$, $z=\rho\cos\varphi$. Utili per solidi con simmetria sferica (palle, coni con vertice nell'origine). Il determinante Jacobiano (calcolo più lungo, che verifichiamo computazionalmente) risulta:

$$\det J = \rho^2\sin\varphi$$

quindi $dV=\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta$.

````python
import sympy as sp

rho, phi, theta = sp.symbols('rho phi theta', positive=True)

x_expr = rho * sp.sin(phi) * sp.cos(theta)
y_expr = rho * sp.sin(phi) * sp.sin(theta)
z_expr = rho * sp.cos(phi)

J = sp.Matrix([
    [sp.diff(x_expr, rho), sp.diff(x_expr, phi), sp.diff(x_expr, theta)],
    [sp.diff(y_expr, rho), sp.diff(y_expr, phi), sp.diff(y_expr, theta)],
    [sp.diff(z_expr, rho), sp.diff(z_expr, phi), sp.diff(z_expr, theta)]
])
det_J = sp.simplify(J.det())
print("Jacobiano coordinate sferiche, determinante:", det_J)

# Volume della sfera di raggio R, tramite integrale triplo in sferiche
R = sp.symbols('R', positive=True)
volume_sfera = sp.integrate(
    sp.integrate(
        sp.integrate(rho**2 * sp.sin(phi), (rho, 0, R)),
        (phi, 0, sp.pi)),
    (theta, 0, 2*sp.pi))
print("\nVolume della sfera di raggio R:", volume_sfera)
print("(confronto: formula nota 4/3 * pi * R^3)")
````

## Esercizio svolto e commentato

**Problema:** calcola il volume del solido $E$ compreso tra il paraboloide $z=x^2+y^2$ e il piano $z=4$.

**Ragionamento:**

**Passo 1 — visualizziamo il solido**: è la regione "sotto" il piano $z=4$ e "sopra" il paraboloide, per $(x,y)$ nel disco dove $x^2+y^2\le4$ (l'intersezione tra paraboloide e piano: $x^2+y^2=4$, una circonferenza di raggio $2$).

**Passo 2 — impostiamo l'integrale triplo**, integrando prima $z$ (tra le due superfici) e poi $(x,y)$ sul disco:

$$V = \iiint_E1\,dV = \iint_{D}\left(\int_{x^2+y^2}^4dz\right)dA = \iint_D(4-x^2-y^2)\,dA$$

dove $D=\{x^2+y^2\le4\}$.

**Passo 3 — passiamo a coordinate polari** (notebook 7.15) per calcolare l'integrale rimanente su $D$: $x^2+y^2=r^2$, e $D$ diventa $\{0\le r\le2,\,0\le\theta\le2\pi\}$:

$$V = \int_0^{2\pi}\int_0^2(4-r^2)\,r\,dr\,d\theta$$

**Passo 4 — integriamo rispetto a $r$**: $\int_0^2(4r-r^3)dr=\left[2r^2-\dfrac{r^4}4\right]_0^2=8-4=4$.

**Passo 5 — integriamo rispetto a $\theta$**: $\int_0^{2\pi}4\,d\theta=8\pi$.

**Conclusione:** $V=8\pi$.

**Perché ridurre prima a un integrale doppio (passo 2), invece di tentare direttamente un integrale triplo in coordinate polari-cilindriche fin dall'inizio, è la strategia più chiara:** il solido ha una struttura naturale — "compreso tra due superfici, sopra una regione piana" — che si presta perfettamente a un primo passo di integrazione rispetto a $z$ (con limiti espliciti tra le due superfici), riducendo il problema tridimensionale a un problema bidimensionale già padroneggiato (notebook 7.16); solo a quel punto conviene passare alle coordinate polari, sfruttando la simmetria circolare della regione $D$ rimanente. Affrontare i problemi per gradi, riducendo la dimensionalità appena possibile, è una strategia generale utile in tutto il calcolo multidimensionale.

````python
import sympy as sp

r, theta = sp.symbols('r theta', real=True)

integrale_r = sp.integrate((4 - r**2) * r, (r, 0, 2))
print("Integrale rispetto a r:", integrale_r)

volume = sp.integrate(integrale_r, (theta, 0, 2 * sp.pi))
print("Volume totale:", volume)
````

## Esercizi da risolvere

2. Usa le coordinate cilindriche per calcolare il volume del cilindro $x^2+y^2\le1$, $0\le z\le3$.
3. Calcola il volume della semisfera $x^2+y^2+z^2\le R^2$, $z\ge0$, usando le coordinate sferiche.
4. *(più impegnativo)* Calcola $\iiint_Ez\,dV$, dove $E$ è la regione compresa tra il cono $z=\sqrt{x^2+y^2}$ e il piano $z=2$ (suggerimento: usa le coordinate cilindriche, con $z$ da $r$ a $2$, e $r$ da $0$ a $2$).

**1.** $\int_0^1\int_0^1\int_0^1z\,dz\,dy\,dx=\int_0^1\int_0^1\dfrac12dy\,dx=\dfrac12\times1\times1=\dfrac12$.

**2.** $V=\int_0^{2\pi}\int_0^1\int_0^3r\,dz\,dr\,d\theta=\int_0^{2\pi}\int_0^13r\,dr\,d\theta=\int_0^{2\pi}3\times\dfrac12d\theta=2\pi\times\dfrac32=3\pi$ (coerente con $\pi r^2h=\pi\times1\times3$).

**3.** $V=\int_0^{2\pi}\int_0^{\pi/2}\int_0^R\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta$ (con $\varphi$ solo da $0$ a $\pi/2$, per restare in $z\ge0$) $=2\pi\times\left[-\cos\varphi\right]_0^{\pi/2}\times\dfrac{R^3}3=2\pi\times1\times\dfrac{R^3}3=\dfrac{2\pi R^3}3$ (metà del volume della sfera completa, coerente).

**4.** In cilindriche, il cono è $z=r$: $V_z=\int_0^{2\pi}\int_0^2\int_r^2z\,r\,dz\,dr\,d\theta$. Integrale interno: $\int_r^2z\,dz=\left[\dfrac{z^2}2\right]_r^2=2-\dfrac{r^2}2$, moltiplicato per $r$: $2r-\dfrac{r^3}2$. $\int_0^2\left(2r-\dfrac{r^3}2\right)dr=\left[r^2-\dfrac{r^4}8\right]_0^2=4-2=2$. Infine $\int_0^{2\pi}2\,d\theta=4\pi$.

## Riepilogo finale

- L'**integrale triplo** $\iiint_Ef\,dV$ generalizza l'integrale doppio a tre dimensioni, con Fubini che riduce a tre integrali unidimensionali successivi; $f=1$ dà il **volume**.
- Le **coordinate cilindriche** ($dV=r\,dr\,d\theta\,dz$) semplificano solidi con simmetria assiale; le **coordinate sferiche** ($dV=\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta$) semplificano solidi con simmetria centrale.
- Una strategia efficace è ridurre prima la dimensionalità (integrando la variabile "verticale" tra due superfici), poi passare a coordinate adatte per l'integrale bidimensionale rimanente.
- Massa e baricentro in tre dimensioni si calcolano con formule dirette generalizzazioni di quelle del notebook 7.16.
