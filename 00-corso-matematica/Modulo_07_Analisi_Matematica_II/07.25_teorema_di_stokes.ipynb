# 7.25 — Teorema di Stokes

**Prerequisiti:** 7.24 Teorema della divergenza, 7.23 Teorema di Green, 7.22 Rotore, 7.19 Integrali di superficie.


## Perché ci interessa

Il teorema di Green (notebook 7.23) collegava una circolazione lungo il bordo di una regione **piana** al rotore su quella regione. Ma cosa succede se la "regione" non è piatta — se è una superficie curva nello spazio, come una cupola o un paraboloide? Il **teorema di Stokes** risponde: la circolazione lungo il bordo di una superficie curva è uguale al flusso del rotore **attraverso** la superficie. È la generalizzazione definitiva del teorema di Green, e il terzo (e ultimo) dei grandi teoremi dell'analisi vettoriale che chiudono questo modulo — tutti manifestazioni dello stesso principio: il teorema fondamentale del calcolo, in versioni via via più generali.

## Teoria, passo dopo passo

### Enunciato del teorema

> **Teorema (Stokes).** Sia $S$ una superficie orientata in $\mathbb R^3$, con bordo $C=\partial S$ una curva chiusa, percorsa in verso coerente con l'orientazione di $S$ (regola della mano destra: se le dita curvano nel verso di percorrenza di $C$, il pollice punta nella direzione della normale $\mathbf n$ scelta per $S$). Se $\mathbf F$ ha componenti di classe $C^1$ su $S$, allora:
> $$\oint_C\mathbf F\cdot d\mathbf r = \iint_S(\operatorname{rot}\mathbf F)\cdot\mathbf n\,dS$$

**Il teorema di Green come caso particolare.** Se $S$ è piatta (contenuta nel piano $xy$, con $\mathbf n=(0,0,1)$), allora $\operatorname{rot}\mathbf F\cdot\mathbf n$ è esattamente la componente $z$ del rotore, cioè $\partial Q/\partial x-\partial P/\partial y$ (notebook 7.22): il teorema di Stokes si riduce esattamente al teorema di Green (notebook 7.23). Stokes è quindi la versione "curva" di Green, valida per superfici qualsiasi, non solo piane.

### Il potere del teorema: cambiare superficie senza cambiare risultato

Una conseguenza notevole: se $S_1$ e $S_2$ sono due superfici **diverse** ma con lo **stesso bordo** $C$ (per esempio, un emisfero e un disco piatto, entrambi con bordo la stessa circonferenza equatoriale), allora:

$$\iint_{S_1}(\operatorname{rot}\mathbf F)\cdot\mathbf n\,dS = \oint_C\mathbf F\cdot d\mathbf r = \iint_{S_2}(\operatorname{rot}\mathbf F)\cdot\mathbf n\,dS$$

Il flusso del rotore **non dipende** dalla forma specifica della superficie, solo dal suo bordo! Questo permette, in pratica, di **scegliere la superficie più comoda** tra tutte quelle con lo stesso bordo — esattamente come, nel notebook 7.20, potevamo scegliere il percorso più comodo per un campo conservativo, sapendo che il risultato non sarebbe cambiato.

### Campi conservativi, di nuovo

Se $\mathbf F=\nabla f$ (conservativo), allora $\operatorname{rot}\mathbf F=\mathbf 0$ ovunque (notebook 7.22): per Stokes, $\oint_C\mathbf F\cdot d\mathbf r=\iint_S\mathbf 0\cdot\mathbf n\,dS=0$, per **ogni** curva chiusa $C$ che sia bordo di una superficie nel dominio di $\mathbf F$ — generalizzando a tre dimensioni la conclusione già raggiunta con Green in due dimensioni (notebook 7.23).

### Esempio: verifica su un emisfero

Consideriamo $\mathbf F(x,y,z)=(-y,x,0)$ (rotazionale, notebook 7.22, con $\operatorname{rot}\mathbf F=(0,0,2)$) e $S$ l'emisfero superiore $x^2+y^2+z^2=1$, $z\ge0$, con bordo $C$ la circonferenza unitaria nel piano $xy$. Per Stokes, la circolazione lungo $C$ deve uguagliare il flusso di $\operatorname{rot}\mathbf F=(0,0,2)$ attraverso $S$ — che, per l'osservazione sul "cambiare superficie", deve coincidere col flusso attraverso il **disco piatto** $z=0$ (stesso bordo!), molto più facile da calcolare: $\iint_{\text{disco}}(0,0,2)\cdot(0,0,1)\,dA=2\times\text{Area(disco)}=2\pi$.

````python
import sympy as sp

t, phi, theta = sp.symbols('t phi theta', real=True)

# Campo rotazionale F=(-y,x,0), rot(F) = (0,0,2)
x, y, z = sp.symbols('x y z', real=True)
F = sp.Matrix([-y, x, 0])
rot_F = sp.Matrix([
    sp.diff(F[2], y) - sp.diff(F[1], z),
    sp.diff(F[0], z) - sp.diff(F[2], x),
    sp.diff(F[1], x) - sp.diff(F[0], y)
])
print("rot(F) =", rot_F.T)

# Circolazione diretta lungo il bordo (circonferenza unitaria nel piano xy)
r_bordo = sp.Matrix([sp.cos(t), sp.sin(t), 0])
r_bordo_prime = r_bordo.diff(t)
F_su_bordo = F.subs({x: r_bordo[0], y: r_bordo[1], z: r_bordo[2]})
circolazione = sp.integrate(sp.simplify((F_su_bordo.T * r_bordo_prime)[0]), (t, 0, 2*sp.pi))
print("\nCircolazione lungo il bordo:", circolazione)

# Flusso del rotore attraverso l'emisfero (parametrizzato in sferiche, raggio 1)
r_emisfero = sp.Matrix([sp.sin(phi)*sp.cos(theta), sp.sin(phi)*sp.sin(theta), sp.cos(phi)])
r_phi = r_emisfero.diff(phi)
r_theta_v = r_emisfero.diff(theta)
normale = r_phi.cross(r_theta_v)

rot_su_superficie = rot_F.subs({x: r_emisfero[0], y: r_emisfero[1], z: r_emisfero[2]})
integranda = sp.trigsimp((rot_su_superficie.T * normale)[0])
flusso = sp.integrate(sp.integrate(integranda, (theta, 0, 2*sp.pi)), (phi, 0, sp.pi/2))
print("Flusso del rotore attraverso l'emisfero:", flusso)

print("\nCoincidono (Stokes verificato)?", sp.simplify(circolazione - flusso) == 0)
````

## Esercizio svolto e commentato

**Problema:** calcola $\oint_C\mathbf F\cdot d\mathbf r$, dove $\mathbf F(x,y,z)=(z,\,x,\,y)$ e $C$ è il bordo del triangolo con vertici $(1,0,0)$, $(0,1,0)$, $(0,0,1)$ (percorso in quest'ordine), usando il teorema di Stokes.

**Ragionamento:**

**Passo 1 — invece di parametrizzare i tre lati del triangolo** (tre integrali curvilinei da sommare), **applichiamo Stokes**, scegliendo come superficie $S$ il triangolo piatto stesso (la scelta più semplice, essendo già il bordo dato il contorno di una superficie piana).

**Passo 2 — calcoliamo il rotore**: $\operatorname{rot}\mathbf F=\left(\dfrac{\partial y}{\partial y}-\dfrac{\partial x}{\partial z},\ \dfrac{\partial z}{\partial z}-\dfrac{\partial y}{\partial x},\ \dfrac{\partial x}{\partial x}-\dfrac{\partial z}{\partial y}\right)=(1-0,\,1-0,\,1-0)=(1,1,1)$.

**Passo 3 — troviamo il vettore normale al triangolo.** Il piano contenente i tre vertici ha equazione $x+y+z=1$ (verificabile: ogni vertice soddisfa questa equazione). Il gradiente di $g(x,y,z)=x+y+z$ è $(1,1,1)$: la normale al piano è proprio $(1,1,1)$ (a meno di normalizzazione), diretta coerentemente con l'orientazione data dall'ordine dei vertici (verificabile con la regola della mano destra, che qui assumiamo coerente).

**Passo 4 — notiamo che $\operatorname{rot}\mathbf F=(1,1,1)$ è parallelo alla normale del piano.** Il flusso attraverso una superficie piatta con normale costante $\mathbf n_0=(1,1,1)/\sqrt3$ (normalizzata) e integranda costante è semplicemente $(\operatorname{rot}\mathbf F)\cdot\mathbf n_0\times\text{Area}(S)=\sqrt3\times\text{Area}(S)$ (essendo $(1,1,1)\cdot(1,1,1)/\sqrt3=3/\sqrt3=\sqrt3$).

**Passo 5 — calcoliamo l'area del triangolo.** I lati sono $(0,1,0)-(1,0,0)=(-1,1,0)$ e $(0,0,1)-(1,0,0)=(-1,0,1)$; l'area è metà della norma del prodotto vettoriale (notebook 6.2, 7.19): $(-1,1,0)\times(-1,0,1)=(1\times1-0\times0,\ 0\times(-1)-(-1)\times1,\ (-1)\times0-1\times(-1))=(1,1,1)$, norma $\sqrt3$, area $=\sqrt3/2$.

**Passo 6 — componiamo**: circolazione $=\sqrt3\times\dfrac{\sqrt3}2=\dfrac32$.

**Conclusione:** $\oint_C\mathbf F\cdot d\mathbf r=\dfrac32$.

**Perché la scelta della superficie **piatta** (invece di una qualsiasi altra superficie con lo stesso bordo triangolare) è quasi sempre la più conveniente per applicare Stokes:** una superficie piatta ha normale **costante**, il che trasforma il flusso del rotore in un semplice prodotto (rotore, anch'esso spesso costante per campi lineari come questo) per l'area — un calcolo enormemente più semplice che parametrizzare una superficie curva arbitraria; il teorema di Stokes garantisce che il risultato è lo stesso qualunque superficie si scelga, quindi conviene sempre cercare, tra tutte le superfici possibili con quel bordo, quella "più piatta" disponibile.

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)
F = sp.Matrix([z, x, y])
rot_F = sp.Matrix([
    sp.diff(F[2], y) - sp.diff(F[1], z),
    sp.diff(F[0], z) - sp.diff(F[2], x),
    sp.diff(F[1], x) - sp.diff(F[0], y)
])
print("rot(F) =", rot_F.T)

v1 = sp.Matrix([-1, 1, 0])
v2 = sp.Matrix([-1, 0, 1])
normale = v1.cross(v2)
print("Normale al piano:", normale.T)

area_triangolo = sp.sqrt((normale.T * normale)[0]) / 2
print("Area del triangolo:", area_triangolo)

flusso = (rot_F.T * normale)[0] / sp.sqrt((normale.T*normale)[0]) * area_triangolo
print("Circolazione (via Stokes):", sp.simplify(flusso))
````

## Riepilogo finale

- Il **teorema di Stokes** $\oint_C\mathbf F\cdot d\mathbf r=\iint_S(\operatorname{rot}\mathbf F)\cdot\mathbf n\,dS$ generalizza il teorema di Green a superfici curve nello spazio; Green ne è il caso particolare con $S$ piatta.
- Il flusso del rotore attraverso una superficie dipende **solo dal suo bordo**: superfici diverse con lo stesso bordo danno lo stesso risultato — permettendo di scegliere sempre la superficie più comoda (spesso quella piatta).
- Per campi conservativi, la circolazione lungo ogni curva chiusa è zero — coerente coi risultati dei notebook 7.20 e 7.23.
- I tre teoremi di questo modulo (Green, divergenza, Stokes) sono tutti espressioni dello stesso principio del teorema fondamentale del calcolo (notebook 5.16): l'integrale di una "derivata" sull'interno è determinato dai valori sul bordo.

Questo conclude il **Modulo 7 (Analisi Matematica II)**: dallo spazio euclideo alle derivate parziali, dai massimi vincolati agli integrali multipli, fino ai tre grandi teoremi dell'analisi vettoriale. Gli strumenti sviluppati qui — gradiente, divergenza, rotore, integrali di linea e di superficie — sono il linguaggio della fisica matematica (elettromagnetismo, fluidodinamica) e riappariranno, formalizzati con ancora maggiore rigore, nel Modulo 15 (Fourier e equazioni alle derivate parziali).
