# 7.11 — Matrice Hessiana

**Prerequisiti:** 7.10 Derivate seconde, 6.23 Forme quadratiche, 5.14 Formula di Taylor.


## Perché ci interessa

Nel notebook 5.14 abbiamo visto che, vicino a un punto, una funzione di una variabile si approssima con un polinomio di Taylor: $f(x_0+h)\approx f(x_0)+f'(x_0)h+\frac12f''(x_0)h^2$. Il termine quadratico, con $f''(x_0)$, ci diceva se la funzione era concava o convessa vicino a quel punto. In più variabili, l'analogo del termine quadratico coinvolge **tutte** le derivate seconde contemporaneamente, organizzate nella matrice Hessiana (anticipata nel notebook 7.10). Questo notebook formalizza la formula di Taylor multidimensionale al secondo ordine, propedeutica alla classificazione di massimi e minimi nel prossimo notebook.

## Teoria, passo dopo passo

### La matrice Hessiana

Per $f:D\to\mathbb R$, $D\subseteq\mathbb R^n$, con derivate parziali seconde continue, la **matrice Hessiana** in un punto $x_0$ è la matrice $n\times n$ (notebook 6.10) delle derivate parziali seconde:

$$H_f(x_0) = \begin{pmatrix}\dfrac{\partial^2f}{\partial x_1^2} & \dfrac{\partial^2f}{\partial x_1\partial x_2} & \cdots\\ \dfrac{\partial^2f}{\partial x_2\partial x_1} & \dfrac{\partial^2f}{\partial x_2^2} & \cdots\\ \vdots & \vdots & \ddots\end{pmatrix}_{x=x_0}$$

Per il teorema di Schwarz (notebook 7.10), $H_f(x_0)$ è **simmetrica**: $\partial^2f/\partial x_i\partial x_j=\partial^2f/\partial x_j\partial x_i$.

### Formula di Taylor al secondo ordine

> **Teorema (Taylor, ordine 2, più variabili).** Se $f$ ha derivate parziali seconde continue vicino a $x_0$, allora per $h\to\mathbf 0$:
> $$f(x_0+h) = f(x_0) + \nabla f(x_0)\cdot h + \frac12\,h^TH_f(x_0)h + o(\|h\|^2)$$

Il termine $h^TH_f(x_0)h$ è esattamente una **forma quadratica** (notebook 6.23) applicata al vettore incremento $h$! Riconosciamo la struttura: il termine di primo ordine $\nabla f(x_0)\cdot h$ (già visto nel notebook 7.8, differenziabilità) è lineare in $h$; il termine di secondo ordine è quadratico, con la matrice simmetrica $H_f(x_0)$ come "coefficienti".

**Perché questa formula generalizza direttamente quella unidimensionale del notebook 5.14**: per $n=1$, $H_f(x_0)=(f''(x_0))$ (una matrice $1\times1$), e $h^TH_f(x_0)h=f''(x_0)h^2$: si ritrova esattamente $f(x_0+h)\approx f(x_0)+f'(x_0)h+\frac12f''(x_0)h^2$.

### Nel punto critico: il termine quadratico domina

Se $x_0$ è un **punto critico** (definiremo formalmente nel prossimo notebook: $\nabla f(x_0)=\mathbf 0$), la formula si semplifica:

$$f(x_0+h) - f(x_0) \approx \frac12\,h^TH_f(x_0)h$$

Il segno di questa quantità — cioè se $f$ cresce o decresce spostandosi da $x_0$ — è determinato **interamente** dal segno della forma quadratica $h^TH_f(x_0)h$: esattamente il tipo di domanda a cui il notebook 6.23 (definita positiva, negativa, indefinita, tramite gli autovalori di $H_f(x_0)$) sapeva già rispondere. Questo è il ponte concettuale che useremo nel prossimo notebook per classificare i punti critici.

### Esempio: la Hessiana di un paraboloide

Per $f(x,y)=x^2+y^2$: $H_f=\begin{pmatrix}2&0\\0&2\end{pmatrix}$ (costante, non dipende dal punto). Gli autovalori sono entrambi $2>0$ (matrice diagonale, notebook 6.17): la forma quadratica è **definita positiva** ovunque — coerente col fatto che il paraboloide $z=x^2+y^2$ è convesso dappertutto, con un unico minimo nell'origine.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
h1, h2 = sp.symbols('h1 h2', real=True)

f = x**2 + y**2
x0, y0 = 1, 1

grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: x0, y: y0})
H = sp.Matrix([[sp.diff(f, x, x), sp.diff(f, x, y)],
               [sp.diff(f, y, x), sp.diff(f, y, y)]]).subs({x: x0, y: y0})
print("Gradiente in (1,1):", grad.T)
print("Hessiana in (1,1):\n", H)

h = sp.Matrix([h1, h2])
# Approssimazione di Taylor al secondo ordine
approx = f.subs({x: x0, y: y0}) + (grad.T * h)[0] + sp.Rational(1,2) * (h.T * H * h)[0]
esatto = f.subs({x: x0 + h1, y: y0 + h2})
print("\nApprossimazione di Taylor (ordine 2):", sp.expand(approx))
print("Valore esatto f(1+h1,1+h2):", sp.expand(esatto))
print("Differenza (deve essere il resto o(||h||^2)):", sp.expand(esatto - approx))
# per un polinomio di grado 2, l'approssimazione al secondo ordine è ESATTA (resto zero)
````

## Esercizio svolto e commentato

**Problema:** calcola la matrice Hessiana di $f(x,y)=x^3-3xy^2$ (nota come "funzione a sella del granchio" o parte reale di $z^3$) in un punto generico, e valutala in $(1,0)$ e in $(0,0)$.

**Ragionamento:**

**Passo 1 — calcoliamo le derivate parziali prime**: $\partial f/\partial x=3x^2-3y^2$, $\partial f/\partial y=-6xy$.

**Passo 2 — calcoliamo le derivate parziali seconde**: $\partial^2f/\partial x^2=6x$. $\partial^2f/\partial x\partial y=-6y$ (derivando $3x^2-3y^2$ rispetto a $y$). $\partial^2f/\partial y\partial x=-6y$ (derivando $-6xy$ rispetto a $x$) — coincide con quella sopra, coerente col teorema di Schwarz. $\partial^2f/\partial y^2=-6x$.

**Passo 3 — scriviamo la matrice Hessiana generale**: $H_f(x,y)=\begin{pmatrix}6x&-6y\\-6y&-6x\end{pmatrix}$.

**Passo 4 — valutiamo in $(1,0)$**: $H_f(1,0)=\begin{pmatrix}6&0\\0&-6\end{pmatrix}$. Essendo diagonale, gli autovalori sono $6$ e $-6$: segni **opposti**, forma quadratica **indefinita** (notebook 6.23).

**Passo 5 — valutiamo in $(0,0)$**: $H_f(0,0)=\begin{pmatrix}0&0\\0&0\end{pmatrix}$: la matrice nulla. Ogni autovalore è $0$: il termine quadratico non fornisce **nessuna** informazione sul segno di $f(h)-f(0,0)$ vicino all'origine (caso "degenere", che richiederà un'analisi più fine, oltre il puro segno della Hessiana).

**Perché il caso $(0,0)$ è particolarmente istruttivo:** mostra che la formula di Taylor al secondo ordine, per quanto potente, non è sempre **decisiva**: quando la Hessiana è nulla (o ha autovalori nulli), il comportamento di $f$ vicino al punto è determinato da termini di ordine superiore (cubici, qui: infatti $f(x,y)=x^3-3xy^2$ **è già** un polinomio omogeneo di terzo grado, quindi vicino a $(0,0)$ il comportamento è dominato proprio dai termini cubici, non quadratici). Questo caso — che riprenderemo nel prossimo notebook — richiede cautela: la Hessiana da sola non basta sempre a classificare un punto critico.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = x**3 - 3*x*y**2

fxx = sp.diff(f, x, x)
fxy = sp.diff(f, x, y)
fyy = sp.diff(f, y, y)

H = sp.Matrix([[fxx, fxy], [fxy, fyy]])
print("Hessiana generale:\n", H)

H_10 = H.subs({x: 1, y: 0})
print("\nHessiana in (1,0):\n", H_10, " autovalori:", H_10.eigenvals())

H_00 = H.subs({x: 0, y: 0})
print("\nHessiana in (0,0):\n", H_00, " autovalori:", H_00.eigenvals())
````

## Riepilogo finale

- La **matrice Hessiana** $H_f$ raccoglie tutte le derivate parziali seconde; per il teorema di Schwarz, è sempre **simmetrica**.
- La **formula di Taylor al secondo ordine**, $f(x_0+h)\approx f(x_0)+\nabla f(x_0)\cdot h+\frac12h^TH_f(x_0)h$, generalizza direttamente il notebook 5.14, con il termine quadratico dato da una forma quadratica (notebook 6.23).
- Nei **punti critici** (dove $\nabla f=\mathbf 0$), il comportamento locale di $f$ è determinato dal segno della forma quadratica associata alla Hessiana.
- Quando la Hessiana è **nulla o degenere**, non fornisce informazioni sufficienti: servono termini di ordine superiore, un caso limite da trattare con cautela.
