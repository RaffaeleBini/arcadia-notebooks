# 7.10 — Derivate seconde

**Prerequisiti:** 7.9 Gradiente, 7.7 Derivate parziali, 5.13 Studio di funzione.


## Perché ci interessa

Nel notebook 5.13, la derivata seconda $f''(x)$ ci diceva se una curva era concava o convessa, informazione essenziale per classificare i punti critici come massimi o minimi. In più variabili, vogliamo lo stesso tipo di informazione — ma ora ci sono **più** derivate seconde possibili: possiamo derivare $\partial f/\partial x$ di nuovo rispetto a $x$, oppure rispetto a $y$; e lo stesso per $\partial f/\partial y$. Questo notebook organizza queste derivate seconde, e dimostra un fatto sorprendente e utilissimo: sotto ipotesi ragionevoli, l'**ordine di derivazione non conta**.

## Teoria, passo dopo passo

### Le quattro derivate parziali seconde (per $n=2$)

Data $f(x,y)$, le derivate parziali $\partial f/\partial x$ e $\partial f/\partial y$ sono a loro volta funzioni di $(x,y)$: possiamo derivarle nuovamente, ottenendo quattro **derivate parziali seconde**:

$$\frac{\partial^2f}{\partial x^2} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial x}\right) \qquad \frac{\partial^2f}{\partial y\,\partial x} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial x}\right)$$
$$\frac{\partial^2f}{\partial x\,\partial y} = \frac{\partial}{\partial x}\left(\frac{\partial f}{\partial y}\right) \qquad \frac{\partial^2f}{\partial y^2} = \frac{\partial}{\partial y}\left(\frac{\partial f}{\partial y}\right)$$

Le due derivate "centrali" $\partial^2f/\partial y\partial x$ e $\partial^2f/\partial x\partial y$ si chiamano **derivate miste**: la notazione va letta da destra a sinistra (prima si deriva rispetto alla variabile più vicina a $f$, poi rispetto all'altra) — attenzione, alcune fonti usano convenzioni diverse, ma qui adottiamo questa, coerente con la notazione $f_{xy}=(f_x)_y$.

### Esempio di calcolo

Per $f(x,y)=x^3y^2$: $\partial f/\partial x=3x^2y^2$, quindi $\partial^2f/\partial x^2=6xy^2$ e $\partial^2f/\partial y\partial x=\partial/\partial y(3x^2y^2)=6x^2y$. D'altra parte, $\partial f/\partial y=2x^3y$, quindi $\partial^2f/\partial x\partial y=\partial/\partial x(2x^3y)=6x^2y$.

**Notiamo che le due derivate miste coincidono**: $\partial^2f/\partial y\partial x=\partial^2f/\partial x\partial y=6x^2y$. Non è un caso isolato.

### Il teorema di Schwarz (uguaglianza delle derivate miste)

> **Teorema (Schwarz).** Se $f$ ha derivate parziali seconde $\partial^2f/\partial x\partial y$ e $\partial^2f/\partial y\partial x$ **continue** in un intorno di $(x_0,y_0)$, allora coincidono in quel punto:
> $$\frac{\partial^2f}{\partial x\,\partial y}(x_0,y_0) = \frac{\partial^2f}{\partial y\,\partial x}(x_0,y_0)$$

**Idea della dimostrazione (schematica).** Si considera il "doppio incremento" $\Delta(h,k)=f(x_0+h,y_0+k)-f(x_0+h,y_0)-f(x_0,y_0+k)+f(x_0,y_0)$, e si dimostra — applicando due volte il teorema di Lagrange (notebook 5.11), una prima rispetto a $x$ e poi rispetto a $y$ (e viceversa, nell'altro ordine) — che $\Delta(h,k)/(hk)$ tende, per $(h,k)\to(0,0)$, sia a $\partial^2f/\partial x\partial y(x_0,y_0)$ sia a $\partial^2f/\partial y\partial x(x_0,y_0)$: essendo lo stesso limite, i due valori devono coincidere.

Per tutte le funzioni "ragionevoli" (polinomi, esponenziali, trigonometriche, e loro composizioni — le stesse che incontriamo sempre in questo corso), le derivate seconde miste sono automaticamente continue, quindi il teorema di Schwarz si applica sempre nella pratica: **l'ordine di derivazione non conta**.

### La matrice Hessiana (anteprima)

Organizzando le quattro derivate seconde in una tabella $2\times2$:

$$H_f = \begin{pmatrix}\partial^2f/\partial x^2 & \partial^2f/\partial x\partial y\\ \partial^2f/\partial y\partial x & \partial^2f/\partial y^2\end{pmatrix}$$

Per il teorema di Schwarz, questa matrice è **simmetrica** (gli elementi fuori diagonale coincidono) — esattamente il tipo di matrice studiato nel notebook 6.23 (forme quadratiche) e nel notebook 6.24 (teorema spettrale)! Questa matrice, chiamata **Hessiana**, sarà lo strumento centrale del prossimo notebook per classificare massimi e minimi.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = x**3 * y**2

fx = sp.diff(f, x)
fy = sp.diff(f, y)
fxx = sp.diff(fx, x)
fxy = sp.diff(fx, y)   # derivata di f_x rispetto a y
fyx = sp.diff(fy, x)   # derivata di f_y rispetto a x
fyy = sp.diff(fy, y)

print("f_x =", fx, " | f_y =", fy)
print("f_xx =", fxx)
print("f_xy (d/dy di f_x) =", fxy)
print("f_yx (d/dx di f_y) =", fyx)
print("f_yy =", fyy)
print("\nTeorema di Schwarz verificato? f_xy == f_yx:", fxy == fyx)

# Matrice Hessiana (simbolica)
H = sp.Matrix([[fxx, fxy], [fyx, fyy]])
print("\nMatrice Hessiana:\n", H)
print("Simmetrica?", H == H.T)
````

## Esercizio svolto e commentato

**Problema:** verifica il teorema di Schwarz per $f(x,y)=\sin(x)e^y+x^2y^3$, calcolando esplicitamente entrambe le derivate miste.

**Ragionamento:**

**Passo 1 — calcoliamo $\partial f/\partial x$**, derivando termine per termine (linearità della derivata, notebook 5.9): $\partial f/\partial x=\cos(x)e^y+2xy^3$.

**Passo 2 — deriviamo $\partial f/\partial x$ rispetto a $y$**, per ottenere $\partial^2f/\partial y\partial x$: $\dfrac{\partial}{\partial y}\left(\cos(x)e^y+2xy^3\right)=\cos(x)e^y+6xy^2$ (il primo termine, $\cos(x)e^y$, ha $\cos(x)$ costante rispetto a $y$, quindi si deriva solo $e^y$; il secondo, $2xy^3$, ha $2x$ costante).

**Passo 3 — calcoliamo $\partial f/\partial y$**: $\partial f/\partial y=\sin(x)e^y+3x^2y^2$.

**Passo 4 — deriviamo $\partial f/\partial y$ rispetto a $x$**, per ottenere $\partial^2f/\partial x\partial y$: $\dfrac{\partial}{\partial x}\left(\sin(x)e^y+3x^2y^2\right)=\cos(x)e^y+6xy^2$.

**Conclusione:** entrambe le derivate miste danno $\cos(x)e^y+6xy^2$: coincidono, coerentemente col teorema di Schwarz (le derivate seconde di questa funzione, somma di prodotti di funzioni elementari, sono ovunque continue).

**Perché, in pratica, non serve mai calcolare entrambe le derivate miste per intero (come abbiamo fatto qui solo per verifica):** grazie al teorema di Schwarz, in tutti gli esercizi successivi (a partire dal prossimo notebook, sulla matrice Hessiana) calcoleremo la derivata mista **una sola volta**, nell'ordine che risulta algebricamente più comodo — per esempio, se $\partial f/\partial x$ ha una forma più semplice di $\partial f/\partial y$, conviene derivare quella rispetto a $y$ piuttosto che il contrario. Il teorema garantisce che il risultato sarebbe comunque lo stesso.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = sp.sin(x) * sp.exp(y) + x**2 * y**3

fxy = sp.diff(f, x, y)  # d^2f/dx dy (SymPy calcola in ordine, ma il risultato è simmetrico)
fyx = sp.diff(f, y, x)
print("d^2f/dx dy =", fxy)
print("d^2f/dy dx =", fyx)
print("Coincidono?", sp.simplify(fxy - fyx) == 0)
````

## Riepilogo finale

- Le **derivate parziali seconde** si ottengono derivando ancora una volta le derivate parziali prime; le derivate **miste** ($\partial^2f/\partial x\partial y$ e $\partial^2f/\partial y\partial x$) sono, a priori, potenzialmente diverse.
- Il **teorema di Schwarz**: se le derivate miste sono continue, coincidono — vero per tutte le funzioni elementari incontrate in questo corso.
- Organizzando le derivate seconde in una tabella, si ottiene la matrice **Hessiana**, automaticamente **simmetrica** grazie al teorema di Schwarz.
- Questa simmetria collega direttamente la Hessiana alla teoria delle forme quadratiche e degli autovalori (notebook 6.23-6.24), che useremo nel prossimo notebook per classificare i punti critici.
