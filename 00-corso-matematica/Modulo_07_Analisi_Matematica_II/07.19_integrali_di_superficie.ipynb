## Perché ci interessa

Nel notebook 7.18 abbiamo imparato a integrare lungo curve. Il passo successivo naturale: integrare su **superfici** — calcolare il flusso di un fluido attraverso una membrana, la carica elettrica totale su una superficie con densità variabile, o semplicemente l'area di una superficie curva. Per farlo ci serve un nuovo strumento vettoriale che non abbiamo ancora incontrato in questo corso: il **prodotto vettoriale**, che produce un vettore perpendicolare a due vettori dati — esattamente ciò che serve per trovare la direzione "normale" (perpendicolare) a una superficie.

## Teoria, passo dopo passo

### Il prodotto vettoriale in $\mathbb R^3$

Dati $\mathbf u=(u_1,u_2,u_3)$ e $\mathbf v=(v_1,v_2,v_3)$, il **prodotto vettoriale** (o prodotto cross) è il vettore:

$$\mathbf u\times\mathbf v = \det\begin{pmatrix}\mathbf i&\mathbf j&\mathbf k\\u_1&u_2&u_3\\v_1&v_2&v_3\end{pmatrix} = (u_2v_3-u_3v_2,\ u_3v_1-u_1v_3,\ u_1v_2-u_2v_1)$$

(un modo mnemonico per ricordare la formula: uno sviluppo di Laplace formale, notebook 6.12, lungo la prima riga di una matrice $3\times3$ con i versori $\mathbf i,\mathbf j,\mathbf k$ della base canonica, notebook 6.8, al posto dei numeri). A differenza del prodotto scalare (che dà un numero), il prodotto vettoriale dà **un vettore**, definito solo in $\mathbb R^3$.

**Proprietà fondamentali** (verificabili direttamente dalla formula):

- $\mathbf u\times\mathbf v$ è **ortogonale** sia a $\mathbf u$ sia a $\mathbf v$ (verificabile calcolando $\mathbf u\cdot(\mathbf u\times\mathbf v)=0$ e $\mathbf v\cdot(\mathbf u\times\mathbf v)=0$, notebook 6.2);
- $\|\mathbf u\times\mathbf v\|=\|\mathbf u\|\|\mathbf v\|\sin\theta$, dove $\theta$ è l'angolo tra $\mathbf u,\mathbf v$ (notebook 6.2): geometricamente, è l'**area** del parallelogramma generato da $\mathbf u,\mathbf v$;
- $\mathbf u\times\mathbf v=-(\mathbf v\times\mathbf u)$ (**anticommutativo**: l'ordine conta, con un cambio di segno);
- $\mathbf u\times\mathbf v=\mathbf 0$ se e solo se $\mathbf u,\mathbf v$ sono paralleli (linearmente dipendenti, notebook 6.6).

### Superfici parametriche

Una **superficie parametrica** in $\mathbb R^3$ è una funzione $\mathbf r(u,v)=(x(u,v),y(u,v),z(u,v))$, con $(u,v)$ variabili in una regione $D\subset\mathbb R^2$ (notebook 7.4) — l'analogo bidimensionale della curva parametrica del notebook 7.18. Le derivate parziali $\mathbf r_u=\partial\mathbf r/\partial u$ e $\mathbf r_v=\partial\mathbf r/\partial v$ (derivate componente per componente, come nel notebook 7.7) sono vettori **tangenti** alla superficie, lungo le direzioni $u$ e $v$.

### Il vettore normale e l'elemento di area

Il vettore $\mathbf r_u\times\mathbf r_v$ è **normale** (perpendicolare) alla superficie nel punto $\mathbf r(u,v)$ (essendo ortogonale a entrambi i vettori tangenti). La sua norma $\|\mathbf r_u\times\mathbf r_v\|$ è l'area del parallelogramma "infinitesimo" generato dagli spostamenti $d\mathbf r_u,d\mathbf r_v$ — l'analogo dello Jacobiano (notebook 7.15), qui espresso tramite il prodotto vettoriale. L'**integrale di superficie** di una funzione scalare $f$ è:

$$\iint_Sf\,dS = \iint_Df(\mathbf r(u,v))\,\|\mathbf r_u\times\mathbf r_v\|\,du\,dv$$

Con $f=1$, si ottiene l'**area della superficie**.

### Esempio: la sfera

Per la sfera di raggio $R$, parametrizzata con le coordinate sferiche (notebook 7.17, con $\rho=R$ fissato): $\mathbf r(\varphi,\theta)=(R\sin\varphi\cos\theta,\,R\sin\varphi\sin\theta,\,R\cos\varphi)$. Calcolando (verifica computazionale sotto): $\|\mathbf r_\varphi\times\mathbf r_\theta\|=R^2\sin\varphi$ — esattamente il fattore $\rho^2\sin\varphi$ (con $\rho=R$ costante) già incontrato nello Jacobiano delle coordinate sferiche nel notebook 7.17: non è un caso, i due calcoli sono strettamente collegati.

````python
import sympy as sp

phi, theta, R = sp.symbols('phi theta R', positive=True)

r = sp.Matrix([R * sp.sin(phi) * sp.cos(theta),
               R * sp.sin(phi) * sp.sin(theta),
               R * sp.cos(phi)])

r_phi = r.diff(phi)
r_theta = r.diff(theta)

# Prodotto vettoriale r_phi x r_theta
normale = r_phi.cross(r_theta)
normale = sp.simplify(normale)
print("Vettore normale r_phi x r_theta:", normale.T)

norma_normale = sp.simplify(sp.sqrt(sp.trigsimp((normale.T * normale)[0])))
print("\n||r_phi x r_theta|| =", norma_normale)

# Area totale della sfera
area = sp.integrate(sp.integrate(norma_normale, (theta, 0, 2*sp.pi)), (phi, 0, sp.pi))
print("\nArea della sfera di raggio R:", area, " (atteso: 4*pi*R^2)")
````

## Esercizio svolto e commentato

**Problema:** calcola l'area della porzione di piano $z=x+y$ che sta sopra il quadrato $D=[0,1]\times[0,1]$ nel piano $xy$.

**Ragionamento:**

**Passo 1 — parametrizziamo la superficie** usando $x,y$ stessi come parametri: $\mathbf r(x,y)=(x,y,x+y)$, con $(x,y)\in D$.

**Passo 2 — calcoliamo i vettori tangenti**: $\mathbf r_x=\partial\mathbf r/\partial x=(1,0,1)$ (derivando ogni componente rispetto a $x$). $\mathbf r_y=\partial\mathbf r/\partial y=(0,1,1)$.

**Passo 3 — calcoliamo il prodotto vettoriale**: $\mathbf r_x\times\mathbf r_y=(0\times1-1\times1,\ 1\times0-1\times1,\ 1\times1-0\times0)=(-1,-1,1)$.

**Passo 4 — calcoliamo la norma**: $\|\mathbf r_x\times\mathbf r_y\|=\sqrt{1+1+1}=\sqrt3$.

**Passo 5 — notiamo che questa norma è costante** (non dipende da $x,y$, essendo il piano "piatto" ovunque): l'integrale si riduce a $\sqrt3$ volte l'area di $D$.

**Conclusione:** $\text{Area}=\iint_D\sqrt3\,dA=\sqrt3\times\text{Area}(D)=\sqrt3\times1=\sqrt3$.

**Perché, per una superficie che è il grafico di una funzione $z=g(x,y)$ (come qui, $g(x,y)=x+y$), esiste una formula scorciatoia che generalizza questo calcolo:** parametrizzando sempre con $\mathbf r(x,y)=(x,y,g(x,y))$, si trova in generale $\mathbf r_x=(1,0,g_x)$, $\mathbf r_y=(0,1,g_y)$, e $\mathbf r_x\times\mathbf r_y=(-g_x,-g_y,1)$, con norma $\sqrt{g_x^2+g_y^2+1}$ — una formula che coinvolge direttamente il **gradiente** di $g$ (notebook 7.9): $\|\mathbf r_x\times\mathbf r_y\|=\sqrt{\|\nabla g\|^2+1}$. Per il piano $g(x,y)=x+y$, $\nabla g=(1,1)$ costante, quindi $\sqrt{1+1+1}=\sqrt3$: coerente col calcolo esplicito, ma ottenibile anche senza rifare il prodotto vettoriale da capo, semplicemente conoscendo il gradiente della funzione che descrive la superficie.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
r = sp.Matrix([x, y, x + y])

rx = r.diff(x)
ry = r.diff(y)
print("r_x =", rx.T, " | r_y =", ry.T)

normale = rx.cross(ry)
print("r_x x r_y =", normale.T)

norma = sp.sqrt((normale.T * normale)[0])
print("Norma:", norma)

area = sp.integrate(sp.integrate(norma, (y, 0, 1)), (x, 0, 1))
print("Area totale:", area)
````

## Riepilogo finale

- Il **prodotto vettoriale** $\mathbf u\times\mathbf v$ (solo in $\mathbb R^3$) produce un vettore ortogonale a entrambi, con norma pari all'area del parallelogramma generato.
- Una **superficie parametrica** $\mathbf r(u,v)$ ha vettore normale $\mathbf r_u\times\mathbf r_v$; la sua norma dà l'elemento di area $dS$.
- L'**integrale di superficie** $\iint_Sf\,dS=\iint_Df(\mathbf r(u,v))\|\mathbf r_u\times\mathbf r_v\|\,du\,dv$ generalizza l'integrale curvilineo del notebook 7.18 a due dimensioni parametriche.
- Per superfici-grafico $z=g(x,y)$, esiste una formula scorciatoia $\|\mathbf r_x\times\mathbf r_y\|=\sqrt{\|\nabla g\|^2+1}$, che collega direttamente l'area superficiale al gradiente.
