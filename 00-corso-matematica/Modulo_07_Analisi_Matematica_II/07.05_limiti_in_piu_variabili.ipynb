# 7.5 — Limiti in più variabili

**Prerequisiti:** 7.4 Funzioni di più variabili, 7.2 Norma e distanza, 5.4 Limiti.


## Perché ci interessa

Nel notebook 5.4 abbiamo definito rigorosamente $\lim_{x\to x_0}f(x)=L$ con la condizione epsilon-delta: per ogni $\varepsilon>0$ esiste $\delta>0$ tale che $0<|x-x_0|<\delta$ implica $|f(x)-L|<\varepsilon$. In una dimensione, "avvicinarsi a $x_0$" significa avvicinarsi **da sinistra o da destra** — solo due direzioni possibili. In $\mathbb R^n$, un punto può essere avvicinato da **infinite** direzioni diverse (o addirittura lungo traiettorie curve, non rettilinee): questo rende il concetto di limite più delicato, e introduce un fenomeno nuovo — un limite può "sembrare" esistere lungo alcune direzioni ma non esistere davvero.

## Teoria, passo dopo passo

### Definizione (epsilon-delta, con la norma)

Sia $f:D\to\mathbb R$, $D\subseteq\mathbb R^n$, e sia $x_0$ un punto tale che ogni palla $B(x_0,r)$ contenga punti di $D$ diversi da $x_0$ (un **punto di accumulazione** di $D$, notebook 7.3). Diciamo che:

$$\lim_{x\to x_0}f(x) = L$$

se per ogni $\varepsilon>0$ esiste $\delta>0$ tale che, per ogni $x\in D$ con $0<\|x-x_0\|<\delta$, si ha $|f(x)-L|<\varepsilon$.

**Confronto diretto con il notebook 5.4**: la definizione è identica, con $|x-x_0|$ (valore assoluto in $\mathbb R$) sostituito da $\|x-x_0\|$ (norma in $\mathbb R^n$, notebook 7.2) — la distanza tra il punto $x$ e $x_0$. La condizione $0<\|x-x_0\|<\delta$ descrive esattamente la palla $B(x_0,\delta)$ **privata** del centro: $x$ deve stare in questa palla forata, e la sua immagine $f(x)$ deve avvicinarsi a $L$ quanto vogliamo.

### La novità: infinite direzioni di avvicinamento

Nel caso unidimensionale, verificare un limite (o la sua non esistenza) richiedeva di controllare solo limite sinistro e destro (notebook 5.4). In $\mathbb R^n$, $x$ può avvicinarsi a $x_0$ lungo **qualsiasi** curva o direzione: rette di ogni pendenza, parabole, spirali... La definizione epsilon-delta richiede che $f(x)\to L$ **indipendentemente** da come $x$ si avvicina — è una condizione molto più forte di "il limite esiste lungo ogni retta".

### Il test delle rette: utile per escludere, non per confermare

Un metodo pratico per **dimostrare che un limite non esiste**: si calcola il limite di $f$ lungo famiglie di rette (o curve) diverse che passano per $x_0$; se si ottengono valori **diversi** a seconda della direzione, il limite (bidimensionale) non può esistere — perché la definizione richiederebbe lo stesso valore $L$ lungo ogni possibile percorso di avvicinamento.

**Attenzione**: il viceversa è **falso**. Trovare lo stesso limite lungo **ogni** retta non basta a garantire che il limite esista: potrebbe non coincidere lungo una traiettoria curva (per esempio una parabola). Vedremo un esempio esplicito di questo fenomeno nell'esercizio svolto.

### Esempio: un limite che non esiste

Consideriamo $f(x,y)=\dfrac{xy}{x^2+y^2}$ per $(x,y)\ne(0,0)$, e chiediamoci se $\lim_{(x,y)\to(0,0)}f(x,y)$ esiste. Avviciniamoci lungo la retta $y=mx$ (con $m$ parametro, notebook 2.2):

$$f(x,mx) = \frac{x(mx)}{x^2+(mx)^2} = \frac{mx^2}{x^2(1+m^2)} = \frac{m}{1+m^2}$$

(semplificando $x^2$, lecito per $x\ne0$). Questo valore **dipende da $m$**: lungo la retta $y=x$ ($m=1$), il limite (lungo quella retta) è $\frac12$; lungo l'asse $x$ ($m=0$), è $0$; lungo $y=2x$ ($m=2$), è $\frac25$. Valori diversi lungo direzioni diverse: il limite bidimensionale **non esiste**.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y, m = sp.symbols('x y m', real=True)
f = (x * y) / (x**2 + y**2)

# Limite lungo la retta y = m*x, per x -> 0
f_lungo_retta = f.subs(y, m * x)
limite_lungo_retta = sp.limit(sp.simplify(f_lungo_retta), x, 0)
print("Limite lungo y=m*x, al variare di m:", limite_lungo_retta)

# Valori numerici per alcuni m
for valore_m in [0, 1, 2, -1]:
    print(f"  m={valore_m}: limite = {limite_lungo_retta.subs(m, valore_m)}")

# Visualizzazione: la funzione non ha un valore limite ben definito vicino all'origine
X = np.linspace(-1, 1, 100)
Y = np.linspace(-1, 1, 100)
X, Y = np.meshgrid(X, Y)
with np.errstate(divide='ignore', invalid='ignore'):
    Z = (X * Y) / (X**2 + Y**2)
Z[np.isnan(Z)] = 0  # valore nell'origine, esclusa dal dominio

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='coolwarm')
ax.set_title("f(x,y) = xy/(x²+y²): nessun limite ben definito in (0,0)")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $\lim_{(x,y)\to(0,0)}f(x,y)$ esiste, per $f(x,y)=\dfrac{x^2y}{x^4+y^2}$.

**Ragionamento:**

**Passo 1 — applichiamo il test delle rette**, lungo $y=mx$: $f(x,mx)=\dfrac{x^2(mx)}{x^4+(mx)^2}=\dfrac{mx^3}{x^4+m^2x^2}=\dfrac{mx^3}{x^2(x^2+m^2)}=\dfrac{mx}{x^2+m^2}$ (semplificando $x^2$). Per $x\to0$: se $m\ne0$, il numeratore tende a $0$ e il denominatore tende a $m^2\ne0$, quindi il limite è $\dfrac0{m^2}=0$. Se $m=0$ (l'asse $x$ stesso, $y=0$): $f(x,0)=0$ per ogni $x\ne0$, limite $0$.

**Passo 2 — il test delle rette dà sempre $0$, per ogni direzione.** Questo **non basta** a concludere che il limite è $0$ (come avvertito nella teoria): dobbiamo controllare anche traiettorie **curve**.

**Passo 3 — proviamo lungo la parabola $y=x^2$** (una scelta naturale, visto che compare $x^4=(x^2)^2$ al denominatore, suggerendo che $y$ "dello stesso ordine" di $x^2$ potrebbe comportarsi diversamente): $f(x,x^2)=\dfrac{x^2\cdot x^2}{x^4+(x^2)^2}=\dfrac{x^4}{x^4+x^4}=\dfrac{x^4}{2x^4}=\dfrac12$ (per $x\ne0$, semplificando $x^4$).

**Passo 4 — confrontiamo.** Lungo la parabola $y=x^2$, il "limite" è $\dfrac12$ per ogni $x\ne0$ (costante, quindi il limite per $x\to0$ lungo questa curva è $\dfrac12$) — **diverso** dal valore $0$ trovato lungo ogni retta.

**Conclusione:** il limite **non esiste** (approcciandosi lungo rette si otterrebbe $0$, ma lungo la parabola $y=x^2$ si ottiene $\frac12$: valori diversi a seconda del percorso).

**Perché il test delle rette, da solo, avrebbe portato a una conclusione sbagliata in questo caso:** questo esempio è il classico controesempio che dimostra la necessità di testare **anche** curve non rettilinee, non solo rette — soprattutto quando la struttura algebrica della funzione (qui, il termine $x^4$ al denominatore) suggerisce che una particolare curva potenzialmente "bilancia" i termini in modo diverso dalle rette. Non esiste una lista finita di curve che garantisca sempre la scoperta della non-esistenza: il test delle rette è solo un primo controllo euristico, mai una dimostrazione di esistenza.

````python
import sympy as sp

x, y, m = sp.symbols('x y m', real=True)
f = (x**2 * y) / (x**4 + y**2)

# Test delle rette
f_retta = sp.simplify(f.subs(y, m * x))
limite_retta = sp.limit(f_retta, x, 0)
print("Limite lungo y=m*x:", limite_retta)

# Test della parabola y = x^2
f_parabola = sp.simplify(f.subs(y, x**2))
print("f lungo y=x^2 (per x != 0):", f_parabola)
limite_parabola = sp.limit(f_parabola, x, 0)
print("Limite lungo y=x^2:", limite_parabola)

print("\nConclusione: limiti diversi (", limite_retta, "vs", limite_parabola, ") -> il limite bidimensionale NON esiste")
````

## Riepilogo finale

- Il limite $\lim_{x\to x_0}f(x)=L$ in $\mathbb R^n$ si definisce come nel notebook 5.4, sostituendo il valore assoluto con la norma (notebook 7.2): richiede che $f(x)\to L$ indipendentemente dalla direzione o traiettoria di avvicinamento a $x_0$.
- Il **test delle rette** permette di dimostrare che un limite **non esiste**, trovando due direzioni con valori diversi — ma non basta mai a **dimostrare** che un limite esiste.
- Un limite può esistere lungo ogni retta ma non esistere in senso bidimensionale, se una curva "nascosta" (come una parabola) dà un valore diverso.
- Per dimostrare rigorosamente l'esistenza di un limite, servono disuguaglianze (come il teorema del confronto) valide in un intorno completo del punto, non solo lungo percorsi specifici.
