# 7.6 — Continuità

**Prerequisiti:** 7.5 Limiti in più variabili, 7.3 Topologia elementare di $\mathbb R^n$, 5.6 Continuità, 5.7 Teoremi fondamentali sulle funzioni continue.


## Perché ci interessa

Nel notebook 5.6, una funzione continua era quella per cui $\lim_{x\to x_0}f(x)=f(x_0)$: nessun "salto" nel grafico. La stessa idea, con il limite multidimensionale appena definito (notebook 7.5), si estende naturalmente a $\mathbb R^n$. Ma la vera ricompensa arriva con i teoremi: il teorema di Weierstrass (notebook 5.7), che garantiva massimo e minimo su un intervallo chiuso e limitato, si estende — con la stessa dimostrazione concettuale — a funzioni continue su insiemi **compatti** (notebook 7.3). Questo risultato sarà essenziale per il notebook 7.12, sulla ricerca di massimi e minimi in più variabili.

## Teoria, passo dopo passo

### Definizione

$f:D\to\mathbb R$ ($D\subseteq\mathbb R^n$) è **continua** in $x_0\in D$ se:

$$\lim_{x\to x_0}f(x) = f(x_0)$$

(se $x_0$ è un punto isolato di $D$, cioè non punto di accumulazione, la continuità vale per convenzione — esattamente come nel caso unidimensionale, notebook 5.6). $f$ è **continua su $D$** se lo è in ogni punto di $D$.

### Le funzioni elementari sono continue

Come nel caso unidimensionale, le operazioni algebriche preservano la continuità: se $f,g$ sono continue in $x_0$, allora $f+g$, $f-g$, $fg$ sono continue in $x_0$, e $f/g$ lo è se $g(x_0)\ne0$ (stessa dimostrazione del notebook 5.6, con la norma al posto del valore assoluto). Le **proiezioni** $\pi_i(x)=x_i$ (che estraggono la $i$-esima coordinata) sono continue — verifica immediata dalla definizione, con $\delta=\varepsilon$: $|x_i-x_{0,i}|\le\|x-x_0\|<\delta=\varepsilon$. Componendo somme, prodotti e proiezioni, ogni funzione **polinomiale** in più variabili (come $f(x,y)=x^2+3xy-y^3$) è continua su tutto $\mathbb R^n$; funzioni razionali sono continue dove il denominatore non si annulla; composizioni con funzioni continue di una variabile (seno, esponenziale, ecc.) restano continue.

### Il legame con il test delle rette

Ricordando il notebook 7.5: se $f$ è continua in $x_0$, allora $f(x)\to f(x_0)$ lungo **ogni** possibile traiettoria di avvicinamento — in particolare lungo ogni retta. Il contrario **non vale**: come visto nell'esempio del notebook 7.5, $\frac{xy}{x^2+y^2}$ non ha nemmeno un limite in $(0,0)$ (fallisce già il test delle rette), quindi non può essere resa continua in quel punto, qualunque valore le si assegni.

### Teorema di Weierstrass (versione multidimensionale)

> **Teorema.** Se $f:K\to\mathbb R$ è continua e $K\subseteq\mathbb R^n$ è **compatto** (chiuso e limitato, notebook 7.3), allora $f$ ammette massimo e minimo assoluti su $K$: esistono $x_M,x_m\in K$ tali che $f(x_m)\le f(x)\le f(x_M)$ per ogni $x\in K$.

Questo teorema generalizza esattamente il notebook 5.7 (dove $K=[a,b]$, l'esempio più semplice di compatto in $\mathbb R$): la dimostrazione (che omettiamo, essendo tecnica) usa lo stesso schema concettuale — successioni in $K$ ammettono sottosuccessioni convergenti (per compattezza, un analogo del teorema di Bolzano-Weierstrass del notebook 4.6), e la continuità di $f$ trasporta la convergenza all'immagine.

**Perché la compattezza è essenziale (e non solo la chiusura o solo la limitatezza)**: $f(x,y)=x$ su $D=\mathbb R^2$ (chiuso, ma non limitato) non ha massimo (cresce indefinitamente); $f(x,y)=\frac1{x^2+y^2}$ sul disco aperto $\{x^2+y^2<1\}$ (limitato, ma non chiuso) non ha massimo (esplode avvicinandosi al bordo escluso). Servono **entrambe** le condizioni insieme.

### Teorema dei valori intermedi (versione multidimensionale)

Il teorema dei valori intermedi (notebook 5.7) si estende a domini **connessi** (concetto che formalizzeremo nel Modulo 12, ma intuitivamente: "fatti di un unico pezzo", senza separazioni): se $f$ è continua su un dominio connesso e assume i valori $a$ e $b$, assume anche ogni valore intermedio.

````python
import sympy as sp
import numpy as np

x, y = sp.symbols('x y', real=True)

# Verifica continuità di una funzione polinomiale per sostituzione diretta del limite
f = x**2 + 3*x*y - y**3
punto = (1, 2)
limite = sp.limit(sp.limit(f, x, punto[0]), y, punto[1])
valore = f.subs({x: punto[0], y: punto[1]})
print("Limite in (1,2):", limite, " | f(1,2):", valore, " | continua?", limite == valore)

# Teorema di Weierstrass: f(x,y) = x^2+y^2 sul disco chiuso di raggio 2 ammette max e min
# Verifica numerica: campioniamo il disco e troviamo max/min approssimati
theta = np.linspace(0, 2*np.pi, 100)
r = np.linspace(0, 2, 50)
R, Theta = np.meshgrid(r, theta)
X, Y = R * np.cos(Theta), R * np.sin(Theta)
Z = X**2 + Y**2
print("\nSu K={x^2+y^2<=4}: min approssimato =", Z.min(), " max approssimato =", Z.max())
print("(min esatto: 0 nell'origine; max esatto: 4 sul bordo, raggio 2)")

# Controesempio: f(x,y)=1/(x^2+y^2) sul disco APERTO non ha massimo (non compatto)
r_vicini_bordo = [0.9, 0.99, 0.999, 0.9999]
for rv in r_vicini_bordo:
    print(f"r={rv}: f = {1/rv**2:.2f}")
print("f cresce senza limite avvicinandosi al bordo escluso: nessun massimo su un dominio non chiuso")
````

## Esercizio svolto e commentato

**Problema:** stabilisci per quali $(x,y)$ la funzione $f(x,y)=\dfrac{\sin(xy)}{x^2+y^2}$ (con $f(0,0)$ da definire) può essere resa continua nell'origine.

**Ragionamento:**

**Passo 1 — calcoliamo il limite lungo le rette $y=mx$**: $f(x,mx)=\dfrac{\sin(mx^2)}{x^2(1+m^2)}$.

**Passo 2 — usiamo il limite notevole $\lim_{t\to0}\frac{\sin t}t=1$ (notebook 5.5)**, con $t=mx^2$ (che tende a $0$ per $x\to0$, se $m\ne0$): riscriviamo $\dfrac{\sin(mx^2)}{x^2(1+m^2)}=\dfrac{\sin(mx^2)}{mx^2}\cdot\dfrac{m}{1+m^2}$. Per $x\to0$: il primo fattore tende a $1$ (limite notevole), quindi il limite lungo la retta è $\dfrac{m}{1+m^2}$.

**Passo 3 — osserviamo che il risultato dipende da $m$**: per esempio, $m=0$ dà limite $0$; $m=1$ dà limite $\frac12$. Il test delle rette dà **valori diversi**.

**Conclusione:** il limite in $(0,0)$ **non esiste** (stesso comportamento della funzione $\frac{xy}{x^2+y^2}$ vista nel notebook 7.5 — non a caso, dato che $\sin(xy)\approx xy$ per $xy$ piccolo, come suggerisce lo sviluppo di Taylor del seno, Modulo 5): quindi **non è possibile** rendere $f$ continua nell'origine, qualunque valore le si assegni in quel punto. $f$ resta continua su $\mathbb R^2\setminus\{(0,0)\}$ (dove è quoziente di funzioni continue con denominatore non nullo), ma non può essere estesa con continuità all'origine.

**Perché riconoscere la "somiglianza" tra $\sin(xy)$ e $xy$ (per $xy$ piccolo) permette di dedurre subito il comportamento di $f$ senza rifare tutti i calcoli da capo:** il limite notevole $\frac{\sin t}t\to1$ ci dice che $\sin(xy)$ "si comporta come" $xy$ vicino all'origine (un'idea che formalizzeremo con gli sviluppi di Taylor); poiché $\frac{xy}{x^2+y^2}$ non ha limite in $(0,0)$ (notebook 7.5), è ragionevole aspettarsi lo stesso comportamento per $\frac{\sin(xy)}{x^2+y^2}$ — un'intuizione che il calcolo esplicito con il test delle rette ha confermato rigorosamente.

````python
import sympy as sp

x, y, m = sp.symbols('x y m', real=True)
f = sp.sin(x * y) / (x**2 + y**2)

f_retta = f.subs(y, m * x)
limite_retta = sp.limit(f_retta, x, 0)
print("Limite lungo y=m*x:", sp.simplify(limite_retta))

for valore_m in [0, 1, 2]:
    print(f"  m={valore_m}: {limite_retta.subs(m, valore_m)}")
````

## Riepilogo finale

- $f$ è **continua** in $x_0$ se $\lim_{x\to x_0}f(x)=f(x_0)$; funzioni polinomiali, razionali (dove definite) e composizioni di funzioni continue restano continue in più variabili.
- Il **teorema di Weierstrass** si estende: una funzione continua su un insieme **compatto** (chiuso e limitato) ammette sempre massimo e minimo assoluti — entrambe le proprietà (chiusura e limitatezza) sono necessarie.
- Se il test delle rette dà valori diversi lungo direzioni diverse, la funzione **non può** essere resa continua in quel punto, qualunque valore le si assegni.
- Questi risultati preparano il terreno per la ricerca sistematica di massimi e minimi (notebook 7.12), che richiederà di sapere **quando** un massimo/minimo esiste prima di cercarlo.
