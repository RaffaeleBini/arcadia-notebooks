# 7.8 — Differenziabilità

**Prerequisiti:** 7.7 Derivate parziali, 6.13 Applicazioni lineari, 5.8 Derivata.


## Perché ci interessa

Nel notebook 7.7 abbiamo visto che l'esistenza delle derivate parziali, da sola, non garantisce nemmeno la continuità: un difetto serio. Il problema è che le derivate parziali guardano solo lungo due direzioni privilegiate (gli assi), ignorando tutte le altre. Serve una nozione più forte, che catturi l'idea — centrale nel notebook 5.8 — che una funzione derivabile "assomiglia a una retta" (la tangente) vicino al punto. In più variabili, l'analogo di "assomigliare a una retta" è "assomigliare a un **piano**" (o, in dimensione ancora più alta, a un'applicazione lineare, notebook 6.13): è il concetto di **differenziabilità**.

## Teoria, passo dopo passo

### Ripensare la derivabilità in una variabile

Nel notebook 5.8, $f$ era derivabile in $x_0$ se il rapporto incrementale aveva limite finito. Riscriviamo questa idea in una forma equivalente, più adatta a generalizzare: $f$ è derivabile in $x_0$ con derivata $f'(x_0)$ se:

$$f(x_0+h) = f(x_0) + f'(x_0)h + o(h) \qquad\text{per } h\to0$$

dove $o(h)$ (si legge "o piccolo di $h$") indica una quantità che tende a $0$ **più velocemente** di $h$: precisamente, $\dfrac{o(h)}h\to0$ per $h\to0$. In parole: vicino a $x_0$, $f$ è ben approssimata dalla retta tangente $f(x_0)+f'(x_0)h$, con un errore trascurabile rispetto a $h$ stesso.

### Definizione di differenziabilità

$f:D\to\mathbb R$, $D\subseteq\mathbb R^n$, è **differenziabile** in $x_0$ (punto interno a $D$) se esiste un vettore $\nabla f(x_0)\in\mathbb R^n$ (che chiameremo **gradiente** nel prossimo notebook) tale che, per $h\in\mathbb R^n$ con $h\to\mathbf 0$:

$$f(x_0+h) = f(x_0) + \nabla f(x_0)\cdot h + o(\|h\|)$$

dove $o(\|h\|)$ soddisfa $\dfrac{o(\|h\|)}{\|h\|}\to0$ per $\|h\|\to0$ (notebook 7.2, per la norma). Il termine $\nabla f(x_0)\cdot h$ (prodotto scalare, notebook 6.20) è **lineare** in $h$: è esattamente il tipo di "applicazione lineare che approssima $f$" richiamato dal notebook 6.13.

### Perché il gradiente, se esiste, coincide con le derivate parziali

Se $f$ è differenziabile con vettore $\nabla f(x_0)=(a_1,\dots,a_n)$, scegliendo $h=(t,0,\dots,0)$ (spostamento lungo il solo asse $x_1$) nella definizione: $f(x_0+te_1)=f(x_0)+a_1t+o(t)$, cioè $\dfrac{f(x_0+te_1)-f(x_0)}t=a_1+\dfrac{o(t)}t\to a_1$ per $t\to0$. Ma il primo membro è proprio la definizione di $\partial f/\partial x_1(x_0)$ (notebook 7.7)! Quindi $a_1=\partial f/\partial x_1(x_0)$, e analogamente per le altre componenti: $\nabla f(x_0)=\left(\dfrac{\partial f}{\partial x_1}(x_0),\dots,\dfrac{\partial f}{\partial x_n}(x_0)\right)$ — le derivate parziali sono **necessarie** per la differenziabilità, ma (come visto nel notebook 7.7) non sono **sufficienti**.

### Differenziabilità implica continuità (finalmente!)

> **Teorema.** Se $f$ è differenziabile in $x_0$, allora $f$ è continua in $x_0$.

**Dimostrazione.** Dalla definizione, $f(x_0+h)-f(x_0)=\nabla f(x_0)\cdot h+o(\|h\|)$. Per $h\to\mathbf 0$: il termine $\nabla f(x_0)\cdot h\to0$ (per Cauchy-Schwarz, notebook 7.2: $|\nabla f(x_0)\cdot h|\le\|\nabla f(x_0)\|\|h\|\to0$), e $o(\|h\|)\to0$ per definizione. Quindi $f(x_0+h)\to f(x_0)$, cioè $f$ è continua in $x_0$. $\blacksquare$

Questo risolve il problema del notebook 7.7: la funzione $\frac{xy}{x^2+y^2}$, pur avendo derivate parziali in $(0,0)$, **non è differenziabile** lì (se lo fosse, sarebbe continua per questo teorema — ma non lo è).

### Un criterio pratico sufficiente

Verificare la differenziabilità direttamente dalla definizione (con il limite dell'$o$ piccolo) è spesso macchinoso. Un criterio pratico, che copre la stragrande maggioranza dei casi:

> Se le derivate parziali $\partial f/\partial x_1,\dots,\partial f/\partial x_n$ esistono e sono **continue** in un intorno di $x_0$, allora $f$ è differenziabile in $x_0$ ($f$ si dice di **classe $C^1$**).

Tutte le funzioni elementari (polinomi, esponenziali, seni, composizioni) hanno derivate parziali continue ovunque siano definite: sono automaticamente differenziabili, senza bisogno di verificarlo caso per caso con la definizione.

````python
import sympy as sp

x, y, h1, h2 = sp.symbols('x y h1 h2', real=True)

# Verifica della formula di differenziabilità per f(x,y) = x^2*y, in (1,2)
f = x**2 * y
x0, y0 = 1, 2

grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: x0, y: y0})
print("Gradiente in (1,2):", grad.T)

# f(x0+h1, y0+h2) - f(x0,y0) - grad . h, diviso per ||h||, deve tendere a 0
h = sp.Matrix([h1, h2])
incremento = f.subs({x: x0 + h1, y: y0 + h2}) - f.subs({x: x0, y: y0})
lineare = (grad.T * h)[0]
resto = sp.expand(incremento - lineare)
print("\nIncremento totale:", sp.expand(incremento))
print("Parte lineare (grad . h):", lineare)
print("Resto o(||h||):", resto)
print("(il resto è quadratico in h1,h2: tende a 0 più velocemente di ||h|| quando h->0)")
````

## Esercizio svolto e commentato

**Problema:** verifica che $f(x,y)=x^2+y^2$ è differenziabile in $(1,1)$, calcolando esplicitamente il resto e mostrando che è $o(\|h\|)$.

**Ragionamento:**

**Passo 1 — calcoliamo il gradiente candidato**: $\partial f/\partial x=2x$, $\partial f/\partial y=2y$; in $(1,1)$: $\nabla f(1,1)=(2,2)$.

**Passo 2 — calcoliamo l'incremento esatto** $f(1+h_1,1+h_2)-f(1,1)$: $(1+h_1)^2+(1+h_2)^2-1-1=1+2h_1+h_1^2+1+2h_2+h_2^2-2=2h_1+2h_2+h_1^2+h_2^2$.

**Passo 3 — sottraiamo la parte lineare** $\nabla f(1,1)\cdot h=2h_1+2h_2$: il resto è $R(h)=(2h_1+2h_2+h_1^2+h_2^2)-(2h_1+2h_2)=h_1^2+h_2^2=\|h\|^2$.

**Passo 4 — verifichiamo che $R(h)=o(\|h\|)$**, cioè che $\dfrac{R(h)}{\|h\|}\to0$ per $\|h\|\to0$: $\dfrac{\|h\|^2}{\|h\|}=\|h\|\to0$. ✓

**Conclusione:** $f(x,y)=x^2+y^2$ è differenziabile in $(1,1)$ (e, per lo stesso calcolo, in ogni punto: il resto è sempre $\|h\|^2$, indipendentemente dal punto di partenza).

**Perché, per un polinomio, il resto $R(h)$ risulta sempre un polinomio di grado $\ge2$ in $h$ (quindi automaticamente $o(\|h\|)$), senza bisogno di invocare il criterio delle derivate parziali continue:** quando si sviluppa $f(x_0+h)$ per un polinomio, i termini di **primo grado** in $h$ si raccolgono esattamente nella parte lineare $\nabla f(x_0)\cdot h$ (è proprio la definizione di derivata parziale, applicata a ciascuna variabile), mentre tutti i termini di grado $\ge2$ in $h$ costituiscono il resto — e ogni tale termine, diviso per $\|h\|$, tende a $0$ (per esempio $h_1^2/\|h\|\le\|h\|\to0$, usando $|h_1|\le\|h\|$). Questo conferma, con un argomento diretto, perché **ogni polinomio è automaticamente differenziabile ovunque** — coerente col criterio generale delle derivate continue, ma visibile qui esplicitamente.

````python
import sympy as sp

h1, h2 = sp.symbols('h1 h2', real=True)

x0, y0 = 1, 1
f = lambda x, y: x**2 + y**2

incremento = f(x0 + h1, y0 + h2) - f(x0, y0)
incremento = sp.expand(incremento)
print("Incremento f(1+h1,1+h2)-f(1,1) =", incremento)

lineare = 2*h1 + 2*h2
resto = sp.expand(incremento - lineare)
print("Resto R(h) =", resto)

norma_h = sp.sqrt(h1**2 + h2**2)
rapporto = resto / norma_h
print("R(h)/||h|| =", sp.simplify(rapporto), "-> tende a 0 quando h->0")
````

## Riepilogo finale

- $f$ è **differenziabile** in $x_0$ se esiste un vettore (il gradiente) tale che $f(x_0+h)=f(x_0)+\nabla f(x_0)\cdot h+o(\|h\|)$: la funzione è ben approssimata da un'applicazione lineare (notebook 6.13) più un resto trascurabile.
- Se $f$ è differenziabile, il vettore $\nabla f(x_0)$ coincide necessariamente con le derivate parziali; ma l'esistenza delle sole derivate parziali **non basta** per la differenziabilità.
- **Differenziabilità implica continuità** (a differenza delle sole derivate parziali, notebook 7.7): risolve la "falla" incontrata nel notebook precedente.
- **Criterio pratico**: derivate parziali continue in un intorno $\Rightarrow$ differenziabilità — soddisfatto da tutte le funzioni elementari.
