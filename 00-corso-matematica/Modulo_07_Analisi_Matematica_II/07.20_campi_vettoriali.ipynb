# 7.20 — Campi vettoriali

**Prerequisiti:** 7.19 Integrali di superficie, 7.9 Gradiente, 7.18 Integrali curvilinei.


## Perché ci interessa

Nei notebook 7.18-7.19 abbiamo usato informalmente l'espressione $\mathbf F(x,y)$ per indicare "una forza che dipende dalla posizione". Questo notebook formalizza il concetto: un **campo vettoriale** assegna un vettore a ogni punto dello spazio — il vento in ogni punto dell'atmosfera, la forza gravitazionale in ogni punto dello spazio, la velocità di un fluido in ogni punto di un tubo. Introduciamo anche il concetto di **campo conservativo**, che generalizza l'osservazione fatta nell'esercizio svolto del notebook 7.18: alcuni campi hanno un "potenziale" da cui derivano, con conseguenze notevoli sul calcolo del lavoro.

## Teoria, passo dopo passo

### Definizione

Un **campo vettoriale** su $D\subseteq\mathbb R^n$ è una funzione $\mathbf F:D\to\mathbb R^n$ che assegna, a ogni punto $x\in D$, un vettore $\mathbf F(x)\in\mathbb R^n$ — è il caso $m=n$ della funzione vettoriale già anticipata nel notebook 7.4. In $\mathbb R^2$: $\mathbf F(x,y)=(P(x,y),Q(x,y))$; in $\mathbb R^3$: $\mathbf F(x,y,z)=(P,Q,R)$, dove $P,Q,R$ sono funzioni scalari, dette **componenti** del campo.

### Campi gradiente

Se $f:D\to\mathbb R$ è una funzione scalare differenziabile (notebook 7.8), il suo **gradiente** $\nabla f$ (notebook 7.9) è esso stesso un campo vettoriale. Un campo $\mathbf F$ che si scrive come $\mathbf F=\nabla f$ per qualche $f$ si dice **campo conservativo** (o campo gradiente), e $f$ si chiama **funzione potenziale** di $\mathbf F$.

**Esempio**: $\mathbf F(x,y)=(x,y)$ (visto nel notebook 7.18) è conservativo, con potenziale $f(x,y)=\frac12(x^2+y^2)$: infatti $\nabla f=(x,y)=\mathbf F$. Il potenziale non è unico: $f(x,y)=\frac12(x^2+y^2)+7$ è ugualmente valido (aggiungere una costante non cambia il gradiente, essendo la derivata di una costante nulla).

### Teorema fondamentale per gli integrali curvilinei

> **Teorema.** Se $\mathbf F=\nabla f$ su $D$, e $\mathbf r(t)$, $t\in[a,b]$, è una curva in $D$ da $\mathbf r(a)=A$ a $\mathbf r(b)=B$, allora:
> $$\int_C\mathbf F\cdot d\mathbf r = f(B)-f(A)$$

**Dimostrazione.** $\int_C\mathbf F\cdot d\mathbf r=\int_a^b\nabla f(\mathbf r(t))\cdot\mathbf r'(t)\,dt$. Ma per la regola della catena in più variabili (la cui idea era già anticipata nel notebook 7.9, teorema del gradiente perpendicolare alle curve di livello, e che qui dimostriamo esplicitamente nel caso che ci serve): $\dfrac{d}{dt}f(\mathbf r(t))=\nabla f(\mathbf r(t))\cdot\mathbf r'(t)$ — l'integranda è **esattamente** la derivata di $f(\mathbf r(t))$ rispetto a $t$! Per il teorema fondamentale del calcolo (notebook 5.16): $\int_a^b\dfrac{d}{dt}f(\mathbf r(t))\,dt=f(\mathbf r(b))-f(\mathbf r(a))=f(B)-f(A)$. $\blacksquare$

Questo generalizza esattamente il teorema fondamentale del calcolo (notebook 5.16) a più variabili: **il lavoro di un campo conservativo dipende solo dai punti estremi**, non dal percorso — esattamente ciò che avevamo osservato empiricamente nell'esercizio del notebook 7.18.

### Come riconoscere un campo conservativo

Verificare direttamente se un campo è un gradiente (cercando esplicitamente il potenziale) può essere macchinoso. Un test necessario, in $\mathbb R^2$: se $\mathbf F=(P,Q)=\nabla f$, allora $P=\partial f/\partial x$ e $Q=\partial f/\partial y$; per il teorema di Schwarz (notebook 7.10), $\partial P/\partial y=\partial^2f/\partial y\partial x=\partial^2f/\partial x\partial y=\partial Q/\partial x$. Quindi:

> **Condizione necessaria**: se $\mathbf F=(P,Q)$ è conservativo, allora $\dfrac{\partial P}{\partial y}=\dfrac{\partial Q}{\partial x}$.

(su domini "semplicemente connessi", senza buchi, questa condizione è anche **sufficiente** — un dettaglio che approfondiremo nel notebook 7.23 col teorema di Green). Se $\partial P/\partial y\ne\partial Q/\partial x$ in qualche punto, il campo **non può** essere conservativo — un test rapido, analogo a quello di derivabilità del notebook 7.7-7.8.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

# Verifica che F=(x,y) è conservativo: cerca il potenziale
P, Q = x, y
condizione = sp.diff(P, y) - sp.diff(Q, x)
print("dP/dy - dQ/dx =", condizione, "-> conservativo (condizione necessaria soddisfatta)")

f_candidato = sp.integrate(P, x)  # prova a integrare P rispetto a x
print("\nPotenziale candidato (integrando P):", f_candidato)
print("Verifica: df/dy =", sp.diff(f_candidato, y), " deve coincidere con Q =", Q)

# Un campo NON conservativo: F=(y,-x) (visto nel notebook 7.18, lavoro non nullo sulla circonferenza)
P2, Q2 = y, -x
condizione2 = sp.diff(P2, y) - sp.diff(Q2, x)
print("\nPer F=(y,-x): dP/dy - dQ/dx =", condizione2, "-> NON conservativo (coerente col lavoro non nullo trovato nel notebook 7.18)")
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $\mathbf F(x,y)=(2xy+y^3,\,x^2+3xy^2)$ è conservativo; se sì, trova un potenziale e usa il teorema fondamentale per calcolare $\int_C\mathbf F\cdot d\mathbf r$ lungo una qualunque curva da $(0,0)$ a $(1,2)$.

**Ragionamento:**

**Passo 1 — applichiamo il test**: $P=2xy+y^3$, $Q=x^2+3xy^2$. $\partial P/\partial y=2x+3y^2$. $\partial Q/\partial x=2x+3y^2$. Coincidono: la condizione necessaria è soddisfatta, il campo **è conservativo** (assumendo il dominio semplicemente connesso, qui tutto $\mathbb R^2$).

**Passo 2 — cerchiamo il potenziale, integrando $P$ rispetto a $x$** (tenendo $y$ costante, l'operazione inversa della derivata parziale, notebook 7.7): $f(x,y)=\int(2xy+y^3)dx=x^2y+xy^3+h(y)$, dove $h(y)$ è una funzione arbitraria di $y$ (il termine "costante" rispetto a $x$, analogo alla costante di integrazione del notebook 5.16, ma qui può dipendere da $y$).

**Passo 3 — determiniamo $h(y)$ imponendo $\partial f/\partial y=Q$**: $\partial f/\partial y=x^2+3xy^2+h'(y)$. Confrontando con $Q=x^2+3xy^2$: deve essere $h'(y)=0$, cioè $h(y)=$ costante (scegliamo $0$, essendo il potenziale definito a meno di una costante).

**Passo 4 — il potenziale è $f(x,y)=x^2y+xy^3$.** Verifichiamo: $\nabla f=(2xy+y^3,\,x^2+3xy^2)=\mathbf F$. ✓

**Passo 5 — applichiamo il teorema fondamentale**: $\int_C\mathbf F\cdot d\mathbf r=f(1,2)-f(0,0)=(1\times2+1\times8)-0=2+8=10$.

**Conclusione:** il lavoro è $10$, **indipendentemente** dalla curva scelta da $(0,0)$ a $(1,2)$ (potrebbe essere una retta, una parabola, una spirale — il risultato non cambia).

**Perché la strategia "integra $P$ rispetto a $x$, poi aggiusta con $h(y)$" funziona sempre, quando il campo è conservativo:** integrando $P=\partial f/\partial x$ rispetto a $x$ si recupera $f$ a meno di un termine che, derivato rispetto a $x$, dà zero — cioè una funzione della sola $y$ (esattamente come, nell'integrazione unidimensionale, la costante di integrazione è "qualunque cosa la cui derivata è zero"). Imponendo poi che la derivata rispetto a $y$ di questa espressione coincida con $Q$ (la seconda componente del campo), si determina $h(y)$ univocamente (a meno di una vera costante numerica) — un metodo sistematico, non un tentativo casuale, per ricostruire il potenziale.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
P = 2*x*y + y**3
Q = x**2 + 3*x*y**2

print("dP/dy - dQ/dx =", sp.diff(P, y) - sp.diff(Q, x))

f = sp.integrate(P, x)
print("Dopo aver integrato P rispetto a x:", f)
# aggiustiamo con h(y) confrontando df/dy con Q
h_prime = sp.simplify(Q - sp.diff(f, y))
print("h'(y) =", h_prime, "(deve essere 0 se il campo è conservativo)")

lavoro = f.subs({x: 1, y: 2}) - f.subs({x: 0, y: 0})
print("\nLavoro da (0,0) a (1,2):", lavoro)
````

## Riepilogo finale

- Un **campo vettoriale** $\mathbf F:D\to\mathbb R^n$ assegna un vettore a ogni punto; un campo **conservativo** è il gradiente di una funzione scalare (il **potenziale**).
- Il **teorema fondamentale per gli integrali curvilinei** generalizza il teorema fondamentale del calcolo (notebook 5.16): per campi conservativi, $\int_C\mathbf F\cdot d\mathbf r=f(B)-f(A)$, indipendentemente dal percorso.
- **Test rapido**: $\partial P/\partial y=\partial Q/\partial x$ è condizione necessaria (e, su domini senza buchi, sufficiente) per la conservatività.
- Il potenziale si ricostruisce integrando una componente e "aggiustando" con l'altra, un metodo sistematico basato sul teorema di Schwarz.
