# 7.15 — Cambiamento di variabili

**Prerequisiti:** 7.14 Integrali multipli, 6.12 Determinanti, 6.14 Matrice associata, 2.13 Coordinate polari.


## Perché ci interessa

Nel notebook 5.17, il metodo di sostituzione trasformava un integrale scomodo in uno più semplice, al costo di moltiplicare per $dx=g'(t)dt$ (la derivata della sostituzione). In più variabili, molte regioni di integrazione (dischi, corone circolari, ellissi) sono scomode da descrivere in coordinate cartesiane $(x,y)$, ma diventano semplicissime in **coordinate polari** (notebook 2.13) o altri sistemi di coordinate. Questo notebook generalizza la sostituzione a più variabili, introducendo il fattore di correzione corretto: lo **Jacobiano**, che misura come un cambiamento di coordinate deforma le aree.

## Teoria, passo dopo passo

### Il problema: le aree si deformano

Una trasformazione di coordinate $(x,y)=T(u,v)$ (per esempio, $x=r\cos\theta,\,y=r\sin\theta$ per le coordinate polari) manda un piccolo rettangolo nel piano $(u,v)$ in una regione **deformata** nel piano $(x,y)$ — non necessariamente un rettangolo, e non necessariamente della stessa area. Per riscrivere correttamente $\iint f(x,y)\,dA$ in termini di $u,v$, dobbiamo tenere conto di **quanto** l'area si deforma localmente in ogni punto.

### Lo Jacobiano

Data la trasformazione $T(u,v)=(x(u,v),y(u,v))$, la sua **matrice Jacobiana** è la matrice delle derivate parziali (esattamente la matrice associata all'approssimazione lineare di $T$ vicino a un punto, notebook 6.14, 7.8, applicata componente per componente):

$$J_T(u,v) = \begin{pmatrix}\dfrac{\partial x}{\partial u} & \dfrac{\partial x}{\partial v}\\[4pt] \dfrac{\partial y}{\partial u} & \dfrac{\partial y}{\partial v}\end{pmatrix}$$

Il suo determinante (notebook 6.12), $\det J_T$, misura il **fattore di scala dell'area** locale: ricordando che $|\det A|$ per una matrice $2\times2$ è l'area del parallelogramma generato dalle colonne (notebook 6.12), $|\det J_T(u_0,v_0)|$ è (localmente, vicino a $(u_0,v_0)$) il rapporto tra l'area della regione trasformata e l'area della regione originale.

### La formula del cambiamento di variabili

> **Teorema.** Se $T:D^*\to D$ è una trasformazione biiettiva e di classe $C^1$ (notebook 7.8) tra regioni di $\mathbb R^2$, allora:
> $$\iint_Df(x,y)\,dA = \iint_{D^*}f(x(u,v),y(u,v))\,|\det J_T(u,v)|\,du\,dv$$

Il fattore $|\det J_T|$ "compensa" esattamente la deformazione dell'area, esattamente come $|g'(t)|$ compensava la deformazione della lunghezza nella sostituzione unidimensionale (notebook 5.17) — non a caso, è la stessa idea, generalizzata da un numero a un determinante.

### Coordinate polari

Per $x=r\cos\theta$, $y=r\sin\theta$ (notebook 2.13):

$$J_T = \begin{pmatrix}\cos\theta & -r\sin\theta\\ \sin\theta & r\cos\theta\end{pmatrix} \qquad \det J_T = r\cos^2\theta+r\sin^2\theta = r(\cos^2\theta+\sin^2\theta) = r$$

(usando l'identità fondamentale della trigonometria, notebook 2.10). Quindi la formula diventa:

$$\iint_Df(x,y)\,dA = \iint_{D^*}f(r\cos\theta,r\sin\theta)\,r\,dr\,d\theta$$

Il fattore $r$ (mai dimenticarlo!) è il prezzo da pagare per usare le coordinate polari: intuitivamente, un piccolo "rettangolo" in $(r,\theta)$ di lati $dr$ e $d\theta$ corrisponde, nel piano $(x,y)$, a un settore quasi-rettangolare di lati $dr$ e $r\,d\theta$ (un arco di lunghezza $r\,d\theta$, per la formula della lunghezza d'arco) — la stessa intuizione geometrica del perché l'area "cresce" proporzionalmente a $r$ man mano che ci si allontana dal centro.

````python
import sympy as sp

r, theta = sp.symbols('r theta', positive=True)

x_expr = r * sp.cos(theta)
y_expr = r * sp.sin(theta)

J = sp.Matrix([[sp.diff(x_expr, r), sp.diff(x_expr, theta)],
               [sp.diff(y_expr, r), sp.diff(y_expr, theta)]])
print("Jacobiano delle coordinate polari:\n", J)
det_J = sp.simplify(J.det())
print("\ndet(J) =", det_J)

# Esempio: area del disco di raggio R, con l'integrale doppio in polari
R = sp.symbols('R', positive=True)
area = sp.integrate(sp.integrate(r, (r, 0, R)), (theta, 0, 2*sp.pi))
print("\nArea del disco di raggio R (via integrale doppio in polari):", area)
print("(confronto: formula nota pi*R^2)")

# Esempio: integrale di una gaussiana su tutto il piano, tecnica classica con coordinate polari
x, y = sp.symbols('x y', real=True)
integrale_gaussiano = sp.integrate(sp.integrate(r * sp.exp(-r**2), (r, 0, sp.oo)), (theta, 0, 2*sp.pi))
print("\nIntegrale di e^(-x^2-y^2) su tutto R^2 (in polari):", integrale_gaussiano)
````

## Esercizio svolto e commentato

**Problema:** calcola $\iint_Dx\,dA$, dove $D$ è il semidisco superiore di raggio $2$: $D=\{x^2+y^2\le4,\,y\ge0\}$.

**Ragionamento:**

**Passo 1 — riconosciamo che $D$ è scomodo in coordinate cartesiane** (i limiti di integrazione coinvolgerebbero radici quadrate), ma è **semplicissimo** in coordinate polari: $D^*=\{0\le r\le2,\ 0\le\theta\le\pi\}$ (un rettangolo nel piano $(r,\theta)$! Il semidisco superiore corrisponde ad angoli da $0$ a $\pi$, notebook 2.13).

**Passo 2 — applichiamo la formula del cambiamento di variabili**, con $f(x,y)=x=r\cos\theta$ e il fattore $r$:

$$\iint_Dx\,dA = \int_0^\pi\int_0^2(r\cos\theta)\,r\,dr\,d\theta = \int_0^\pi\int_0^2r^2\cos\theta\,dr\,d\theta$$

**Passo 3 — integriamo prima rispetto a $r$** (tenendo $\theta$ fisso): $\int_0^2r^2\cos\theta\,dr=\cos\theta\left[\dfrac{r^3}3\right]_0^2=\cos\theta\times\dfrac83$.

**Passo 4 — integriamo rispetto a $\theta$**: $\int_0^\pi\dfrac83\cos\theta\,d\theta=\dfrac83\left[\sin\theta\right]_0^\pi=\dfrac83(\sin\pi-\sin0)=\dfrac83(0-0)=0$.

**Conclusione:** $\iint_Dx\,dA=0$.

**Perché questo risultato ha senso senza nemmeno fare il calcolo, per simmetria:** il semidisco superiore è **simmetrico** rispetto all'asse $y$ (riflettendo $x\to-x$, la regione resta invariata), mentre l'integranda $f(x,y)=x$ è **dispari** rispetto a questa riflessione ($f(-x,y)=-f(x,y)$). Per ogni punto con coordinata $x$ positiva, c'è un punto speculare con coordinata $-x$ e lo stesso peso: i loro contributi si cancellano esattamente a coppie, dando somma totale zero — lo stesso principio di simmetria già visto per gli integrali di funzioni dispari su intervalli simmetrici (Modulo 5), qui applicato a una regione simmetrica del piano. Riconoscere questa simmetria **prima** di calcolare avrebbe permesso di prevedere il risultato $0$ senza fare nessun conto.

````python
import sympy as sp

r, theta = sp.symbols('r theta', real=True)

integrale_interno = sp.integrate(r**2 * sp.cos(theta), (r, 0, 2))
print("Integrale interno (rispetto a r):", integrale_interno)

integrale_totale = sp.integrate(integrale_interno, (theta, 0, sp.pi))
print("Integrale doppio totale:", integrale_totale)
````

## Esercizi da risolvere

1. Calcola $\iint_Dy\,dA$ sullo stesso semidisco $D=\{x^2+y^2\le4,\,y\ge0\}$ (qui non c'è simmetria che annulla il risultato: calcola esplicitamente).
2. Usa le coordinate polari per calcolare $\iint_D(x^2+y^2)\,dA$, dove $D$ è il disco unitario $x^2+y^2\le1$.
3. Calcola l'area della corona circolare $1\le x^2+y^2\le4$ usando un integrale doppio in coordinate polari.
4. *(più impegnativo)* Verifica che il famoso integrale gaussiano $\int_{-\infty}^{+\infty}e^{-x^2}dx=\sqrt\pi$ segue dal calcolo (fatto sopra) di $\iint_{\mathbb R^2}e^{-x^2-y^2}dA$ in coordinate polari, osservando che l'integrale doppio si **fattorizza** come prodotto di due integrali unidimensionali identici (suggerimento: $e^{-x^2-y^2}=e^{-x^2}e^{-y^2}$, e usa il teorema di Fubini su un rettangolo che tende a $\mathbb R^2$).

**1.** $\iint_Dy\,dA=\int_0^\pi\int_0^2(r\sin\theta)r\,dr\,d\theta=\int_0^\pi\sin\theta\left[\dfrac{r^3}3\right]_0^2d\theta=\dfrac83\int_0^\pi\sin\theta\,d\theta=\dfrac83\left[-\cos\theta\right]_0^\pi=\dfrac83(1-(-1))=\dfrac{16}3$.

**2.** In polari, $x^2+y^2=r^2$: $\iint_Dr^2\cdot r\,dr\,d\theta=\int_0^{2\pi}\int_0^1r^3\,dr\,d\theta=2\pi\times\left[\dfrac{r^4}4\right]_0^1=2\pi\times\dfrac14=\dfrac\pi2$.

**3.** $\text{Area}=\int_0^{2\pi}\int_1^2r\,dr\,d\theta=2\pi\times\left[\dfrac{r^2}2\right]_1^2=2\pi\times\dfrac{4-1}2=2\pi\times\dfrac32=3\pi$ (coerente con la formula elementare: area del cerchio grande meno area del cerchio piccolo, $4\pi-\pi=3\pi$).

**4.** $\iint_{\mathbb R^2}e^{-x^2-y^2}dA=\pi$ (dal calcolo in polari mostrato sopra nel notebook, con $R\to\infty$). D'altra parte, per Fubini: $\iint_{\mathbb R^2}e^{-x^2}e^{-y^2}dxdy=\left(\int_{-\infty}^{+\infty}e^{-x^2}dx\right)\left(\int_{-\infty}^{+\infty}e^{-y^2}dy\right)=\left(\int_{-\infty}^{+\infty}e^{-x^2}dx\right)^2$ (i due integrali sono identici, avendo la stessa forma). Uguagliando: $\left(\int_{-\infty}^{+\infty}e^{-x^2}dx\right)^2=\pi$, quindi $\int_{-\infty}^{+\infty}e^{-x^2}dx=\sqrt\pi$ (la radice positiva, essendo l'integranda sempre positiva).

## Riepilogo finale

- Un cambiamento di coordinate $T(u,v)=(x,y)$ deforma localmente le aree secondo il fattore $|\det J_T|$, il valore assoluto del determinante della **matrice Jacobiana**.
- La **formula del cambiamento di variabili** $\iint_Df\,dA=\iint_{D^*}f(T(u,v))\,|\det J_T|\,du\,dv$ generalizza la sostituzione del notebook 5.17.
- Per le **coordinate polari**, $\det J_T=r$: il fattore $r\,dr\,d\theta$ è essenziale, mai da dimenticare.
- Le coordinate polari trasformano dischi, settori e corone circolari in **rettangoli**, semplificando enormemente molti integrali — incluso il celebre integrale gaussiano.
