## Teoria, passo dopo passo

### Regioni di tipo I e tipo II

Una regione $D$ è di **tipo I** se si descrive come $D=\{a\le x\le b,\ \phi_1(x)\le y\le\phi_2(x)\}$ (compresa tra due curve, al variare di $x$) — la forma usata nel notebook 7.14. È di **tipo II** se si descrive come $D=\{c\le y\le d,\ \psi_1(y)\le x\le\psi_2(y)\}$ (compresa tra due curve, al variare di $y$). Molte regioni sono **di entrambi i tipi**, e si può scegliere quale descrizione usare — una scelta che, come visto nell'esercizio 4 del notebook 7.14, può rendere un integrale altrimenti impossibile (con primitive non elementari) perfettamente calcolabile.

**Come decidere quale ordine usare**: si prova a immaginare "affettare" la regione con rette verticali (tipo I, integrando prima rispetto a $y$) oppure orizzontali (tipo II, integrando prima rispetto a $x$); si sceglie la descrizione che dà i limiti di integrazione più semplici, o che rende l'integranda risolvibile.

### Massa di una lamina piana

Se $D$ rappresenta una lamina piana con **densità** $\rho(x,y)$ (massa per unità di area, non necessariamente uniforme), la massa totale è:

$$M = \iint_D\rho(x,y)\,dA$$

(esattamente come, in una dimensione, la massa di un'asta con densità lineare $\rho(x)$ era $\int_a^b\rho(x)dx$ — un'applicazione diretta dell'integrale come "somma continua" già vista nel Modulo 5).

### Baricentro

Il **baricentro** (centro di massa) $(\bar x,\bar y)$ di una lamina con densità $\rho$ è:

$$\bar x = \frac1M\iint_Dx\,\rho(x,y)\,dA \qquad \bar y = \frac1M\iint_Dy\,\rho(x,y)\,dA$$

Se $\rho$ è costante (lamina omogenea), $\rho$ si semplifica in entrambe le formule (compare sia al numeratore sia in $M$), e $(\bar x,\bar y)$ dipende solo dalla **forma geometrica** di $D$, non dal materiale: il baricentro geometrico.

### Valore medio di una funzione

Generalizzando il valore medio del notebook 5.16 (teorema della media integrale): il **valore medio** di $f$ su una regione $D$ di area $A(D)=\iint_D1\,dA$ è:

$$\bar f = \frac1{A(D)}\iint_Df(x,y)\,dA$$

Interpretazione: è "l'altezza costante" che, distribuita uniformemente su $D$, darebbe lo stesso volume sotto il grafico di $f$.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

# Baricentro di un triangolo omogeneo: D = {0<=x<=1, 0<=y<=1-x}
rho = 1  # densità costante (baricentro geometrico)
M = sp.integrate(sp.integrate(rho, (y, 0, 1 - x)), (x, 0, 1))
print("Massa (= Area, densità unitaria):", M)

Mx_bar = sp.integrate(sp.integrate(x * rho, (y, 0, 1 - x)), (x, 0, 1))
My_bar = sp.integrate(sp.integrate(y * rho, (y, 0, 1 - x)), (x, 0, 1))
x_bar = Mx_bar / M
y_bar = My_bar / M
print("Baricentro (x_bar, y_bar):", (x_bar, y_bar))
print("(atteso: il baricentro di un triangolo è a 1/3 da ciascun lato)")

# Valore medio di f(x,y) = x^2+y^2 sul quadrato [0,1]x[0,1]
f = x**2 + y**2
area = 1
media = sp.integrate(sp.integrate(f, (y, 0, 1)), (x, 0, 1)) / area
print("\nValore medio di x^2+y^2 su [0,1]x[0,1]:", media)
````

## Esercizio svolto e commentato

**Problema:** una lamina triangolare con vertici $(0,0)$, $(2,0)$, $(0,2)$ ha densità $\rho(x,y)=x+y$. Trova la massa totale e la coordinata $\bar x$ del baricentro.

**Ragionamento:**

**Passo 1 — descriviamo la regione** (tipo I): il triangolo ha ipotenusa lungo la retta $x+y=2$; per $0\le x\le2$, $y$ varia da $0$ a $2-x$. $D=\{0\le x\le2,\ 0\le y\le2-x\}$.

**Passo 2 — calcoliamo la massa**: $M=\int_0^2\int_0^{2-x}(x+y)\,dy\,dx$. Integrale interno: $\int_0^{2-x}(x+y)dy=\left[xy+\dfrac{y^2}2\right]_0^{2-x}=x(2-x)+\dfrac{(2-x)^2}2$.

**Passo 3 — sviluppiamo l'espressione** (ponendo $t=2-x$ per semplicità, poi sostituendo): $x(2-x)+\dfrac{(2-x)^2}2=(2-x)\left(x+\dfrac{2-x}2\right)=(2-x)\times\dfrac{2x+2-x}2=(2-x)\times\dfrac{x+2}2=\dfrac{(2-x)(2+x)}2=\dfrac{4-x^2}2$ (differenza di quadrati, notebook 1.2).

**Passo 4 — integriamo rispetto a $x$**: $M=\int_0^2\dfrac{4-x^2}2dx=\dfrac12\left[4x-\dfrac{x^3}3\right]_0^2=\dfrac12\left(8-\dfrac83\right)=\dfrac12\times\dfrac{16}3=\dfrac83$.

**Passo 5 — calcoliamo il numeratore per $\bar x$**: $\iint_Dx(x+y)dA=\int_0^2x\times\dfrac{4-x^2}2dx$ (riusando il risultato del passo 3, moltiplicato per $x$) $=\dfrac12\int_0^2(4x-x^3)dx=\dfrac12\left[2x^2-\dfrac{x^4}4\right]_0^2=\dfrac12\left(8-4\right)=2$.

**Passo 6 — dividiamo per la massa**: $\bar x=\dfrac2{8/3}=2\times\dfrac38=\dfrac34$.

**Conclusione:** la massa totale è $\dfrac83$, e $\bar x=\dfrac34$.

**Perché riutilizzare il risultato dell'integrale interno del passo 3 (invece di ricalcolarlo da capo per $\bar x$) è un'ottimizzazione naturale, non solo una scorciatoia di comodo:** l'integrale interno rispetto a $y$, calcolato una volta per $M$, ricompare — moltiplicato per un fattore $x$ — nel calcolo di $\bar x$; riconoscere questa struttura comune permette di risparmiare un calcolo intero, un'abitudine utile ogni volta che si devono calcolare più quantità correlate (massa, baricentro, momenti) sulla stessa regione.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
rho = x + y

M = sp.integrate(sp.integrate(rho, (y, 0, 2 - x)), (x, 0, 2))
print("Massa:", M)

Mx = sp.integrate(sp.integrate(x * rho, (y, 0, 2 - x)), (x, 0, 2))
x_bar = Mx / M
print("x_bar:", x_bar)
````

## Riepilogo finale

- Una regione può essere di **tipo I** (limiti $y$ in funzione di $x$) o **tipo II** (limiti $x$ in funzione di $y$); scegliere la descrizione giusta può semplificare drasticamente un integrale.
- La **massa** di una lamina con densità $\rho$ è $\iint_D\rho\,dA$; il **baricentro** $(\bar x,\bar y)$ pesa le coordinate con la densità.
- Il **valore medio** di $f$ su $D$ è $\frac1{A(D)}\iint_Df\,dA$, generalizzando il teorema della media integrale del notebook 5.16.
- Riconoscere strutture comuni tra calcoli correlati (come nell'esercizio svolto) permette di risparmiare lavoro.
