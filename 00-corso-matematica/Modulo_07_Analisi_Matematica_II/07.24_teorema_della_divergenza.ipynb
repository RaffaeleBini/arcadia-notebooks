# 7.24 — Teorema della divergenza

**Prerequisiti:** 7.23 Teorema di Green, 7.21 Divergenza, 7.19 Integrali di superficie, 7.17 Integrali tripli.


## Perché ci interessa

Il teorema di Green (notebook 7.23) collegava una circolazione lungo il bordo di una regione **piana** a un integrale doppio sulla regione. Il **teorema della divergenza** (o teorema di Gauss) è l'analogo in tre dimensioni: collega il **flusso** di un campo attraverso la superficie che racchiude un solido a un integrale triplo della divergenza sul solido stesso. È lo strumento che rende rigorosa l'intuizione fisica della divergenza vista nel notebook 7.21 (sorgenti e pozzi), ed è alla base delle equazioni di Maxwell dell'elettromagnetismo e delle leggi di conservazione in fisica.

## Teoria, passo dopo passo

### Il flusso attraverso una superficie

Dato un campo vettoriale $\mathbf F$ e una superficie $S$ (parametrizzata da $\mathbf r(u,v)$, notebook 7.19) con vettore normale $\mathbf n$, il **flusso** di $\mathbf F$ attraverso $S$ è:

$$\iint_S\mathbf F\cdot\mathbf n\,dS = \iint_D\mathbf F(\mathbf r(u,v))\cdot(\mathbf r_u\times\mathbf r_v)\,du\,dv$$

(usando il vettore normale $\mathbf r_u\times\mathbf r_v$ del notebook 7.19, non normalizzato: la sua norma "assorbe" già l'elemento di area, quindi non serve dividere). Fisicamente: se $\mathbf F$ è la velocità di un fluido, il flusso misura **quanto fluido attraversa $S$ per unità di tempo**, contato positivamente se nella direzione di $\mathbf n$.

### Superfici chiuse e normale uscente

Una superficie **chiusa** (come una sfera o la superficie di un cubo) racchiude un solido $E$. Per convenzione, si orienta $S=\partial E$ con la normale **uscente** (che punta verso l'esterno del solido).

### Enunciato del teorema

> **Teorema (della divergenza, o di Gauss).** Sia $E\subset\mathbb R^3$ un solido delimitato da una superficie chiusa $S=\partial E$, con normale uscente. Se $\mathbf F$ ha componenti di classe $C^1$ su $E$, allora:
> $$\iint_S\mathbf F\cdot\mathbf n\,dS = \iiint_E\operatorname{div}\mathbf F\,dV$$

**Perché questo risultato è l'analogo naturale del teorema di Green**: entrambi affermano che "l'accumulo di una grandezza derivata (rotore o divergenza) sull'interno è determinato dal comportamento sul bordo (circolazione o flusso)" — lo stesso principio del teorema fondamentale del calcolo (notebook 5.16), applicato a dimensioni successive: da un intervallo con bordo di 2 punti, a una regione piana con bordo una curva, a un solido con bordo una superficie.

### Interpretazione: sorgenti totali = flusso uscente

Il teorema formalizza l'intuizione del notebook 7.21: il flusso netto uscente da una superficie chiusa è uguale alla somma (integrale) di tutte le "sorgenti" e "pozzi" (divergenza) racchiusi all'interno. Se $\operatorname{div}\mathbf F=0$ ovunque in $E$ (campo solenoidale, notebook 7.21), il flusso attraverso **qualsiasi** superficie chiusa che racchiude solo punti di $E$ è **zero**: tutto ciò che entra deve uscire, non essendoci sorgenti né pozzi interni.

### Esempio: il campo radiale attraverso la sfera

Per $\mathbf F(x,y,z)=(x,y,z)$ (campo radiale, generalizzazione tridimensionale del campo del notebook 7.21): $\operatorname{div}\mathbf F=1+1+1=3$ (costante). Per una sfera $E$ di raggio $R$: $\iiint_E3\,dV=3\times\text{Volume}(E)=3\times\dfrac43\pi R^3=4\pi R^3$. Per il teorema, il flusso attraverso la superficie sferica è esattamente $4\pi R^3$ — senza dover calcolare direttamente un integrale di superficie (che pure daremmo lo stesso risultato, come verifichiamo).

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)

# Campo radiale F=(x,y,z), divergenza
F = sp.Matrix([x, y, z])
div_F = sum(sp.diff(F[i], v) for i, v in enumerate([x, y, z]))
print("div(F) =", div_F)

# Integrale triplo della divergenza sulla sfera di raggio R
R = sp.symbols('R', positive=True)
rho, phi, theta = sp.symbols('rho phi theta', positive=True)
integrale_triplo = sp.integrate(
    sp.integrate(
        sp.integrate(div_F * rho**2 * sp.sin(phi), (rho, 0, R)),
        (phi, 0, sp.pi)),
    (theta, 0, 2*sp.pi))
print("Integrale triplo di div(F) sulla sfera:", integrale_triplo)

# Verifica diretta: flusso attraverso la superficie sferica
r_sup = sp.Matrix([R*sp.sin(phi)*sp.cos(theta), R*sp.sin(phi)*sp.sin(theta), R*sp.cos(phi)])
r_phi = r_sup.diff(phi)
r_theta = r_sup.diff(theta)
normale = r_phi.cross(r_theta)  # non normalizzato

F_su_superficie = F.subs({x: r_sup[0], y: r_sup[1], z: r_sup[2]})
integranda_flusso = sp.trigsimp((F_su_superficie.T * normale)[0])
print("\nIntegranda del flusso:", integranda_flusso)

flusso = sp.integrate(sp.integrate(integranda_flusso, (theta, 0, 2*sp.pi)), (phi, 0, sp.pi))
print("Flusso attraverso la superficie sferica:", flusso)
print("\nCoincidono (teorema della divergenza verificato)?", sp.simplify(integrale_triplo - flusso) == 0)
````

## Esercizio svolto e commentato

**Problema:** calcola il flusso di $\mathbf F(x,y,z)=(x^3,\,y^3,\,z^3)$ attraverso la superficie del cubo $[0,1]^3$, usando il teorema della divergenza.

**Ragionamento:**

**Passo 1 — invece di calcolare il flusso direttamente** (che richiederebbe di sommare i contributi su tutte e sei le facce del cubo, con orientazioni diverse), **applichiamo il teorema della divergenza**.

**Passo 2 — calcoliamo la divergenza**: $\operatorname{div}\mathbf F=\partial(x^3)/\partial x+\partial(y^3)/\partial y+\partial(z^3)/\partial z=3x^2+3y^2+3z^2$.

**Passo 3 — impostiamo l'integrale triplo** sul cubo $E=[0,1]^3$:

$$\iiint_E(3x^2+3y^2+3z^2)\,dV$$

**Passo 4 — sfruttiamo la simmetria**: per la simmetria del cubo rispetto a $x,y,z$ (lo stesso ruolo per ciascuna variabile), i tre integrali $\iiint_E3x^2\,dV$, $\iiint_E3y^2\,dV$, $\iiint_E3z^2\,dV$ sono **identici** (basta rinominare le variabili di integrazione). Calcoliamo uno solo: $\iiint_E3x^2\,dV=3\int_0^1x^2dx\times\int_0^11\,dy\times\int_0^11\,dz=3\times\dfrac13\times1\times1=1$.

**Passo 5 — sommiamo i tre contributi identici**: $1+1+1=3$.

**Conclusione:** il flusso è $3$.

**Perché riconoscere la simmetria (passo 4) evita di dover calcolare tre integrali tripli separati:** quando l'integranda e il dominio hanno la stessa struttura rispetto a più variabili (qui, il cubo è invariante scambiando $x,y,z$, e l'integranda ha la stessa forma $3x^2+3y^2+3z^2$ in ciascuna), i singoli contributi devono necessariamente essere uguali — un principio di simmetria che vale ben oltre questo esempio specifico, e che conviene sempre cercare prima di lanciarsi in calcoli espliciti e ripetitivi.

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)
div_F = 3*x**2 + 3*y**2 + 3*z**2

flusso = sp.integrate(sp.integrate(sp.integrate(div_F, (x, 0, 1)), (y, 0, 1)), (z, 0, 1))
print("Flusso (via teorema della divergenza):", flusso)
````

## Riepilogo finale

- Il **flusso** $\iint_S\mathbf F\cdot\mathbf n\,dS$ misura quanto un campo "attraversa" una superficie.
- Il **teorema della divergenza**: $\iint_S\mathbf F\cdot\mathbf n\,dS=\iiint_E\operatorname{div}\mathbf F\,dV$, per $S=\partial E$ con normale uscente — l'analogo tridimensionale del teorema di Green.
- Per campi **solenoidali** ($\operatorname{div}\mathbf F=0$), il flusso attraverso **qualsiasi** superficie chiusa è sempre nullo.
- Sfruttare simmetrie nell'integranda e nel dominio (come nell'esercizio svolto) può semplificare notevolmente il calcolo dell'integrale triplo risultante.
