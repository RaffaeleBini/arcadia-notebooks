## Perché ci interessa

Nel Modulo 5 abbiamo studiato funzioni di **una sola** variabile: $f(x)$, dove $x$ è un numero reale. Ma la maggior parte dei fenomeni reali dipende da più grandezze contemporaneamente: la temperatura di una stanza dipende da tre coordinate spaziali (e dal tempo); il profitto di un'azienda dipende dal prezzo e dalla quantità prodotta; l'altitudine di un punto su una mappa dipende dalla latitudine e dalla longitudine. Per studiare rigorosamente funzioni che dipendono da più variabili, dobbiamo prima capire bene **l'ambiente** in cui vivono: lo spazio $\mathbb R^n$, dotato non solo della struttura di spazio vettoriale (Modulo 6) ma anche della possibilità di misurare distanze e angoli. Questo modulo, che ricordiamo essere pensato per essere studiato in parallelo al Modulo 5, applica sistematicamente gli strumenti dell'analisi rigorosa (limiti, continuità, derivate, integrali) e dell'algebra lineare (vettori, matrici, autovalori) a un ambiente a più dimensioni.

## Teoria, passo dopo passo

### $\mathbb R^n$ come insieme di punti

$\mathbb R^n$ è l'insieme delle $n$-uple ordinate di numeri reali: $x=(x_1,x_2,\dots,x_n)$, con $x_i\in\mathbb R$ (notebook 5.1). Per $n=1$ è la retta reale (Modulo 5); per $n=2$ il piano cartesiano (notebook 2.1); per $n=3$ lo spazio tridimensionale ordinario. Per $n>3$ non abbiamo più un'intuizione visiva diretta, ma la struttura algebrica resta identica — è proprio la forza dell'astrazione algebrica (già sperimentata nel notebook 6.3) che ci permette di ragionare rigorosamente anche "oltre" l'intuizione geometrica.

### La doppia natura di $\mathbb R^n$: punti e vettori

In questo modulo useremo $\mathbb R^n$ in due modi intercambiabili, esattamente come già capitava nel Modulo 6 (notebook 6.1): come insieme di **punti** (posizioni nello spazio, es. il punto dove si misura la temperatura) e come **spazio vettoriale** (notebook 6.3), dove ogni punto può anche essere pensato come il vettore che va dall'origine a quel punto. Questa doppia lettura ci permette di applicare somma, prodotto per scalare, e — come vedremo — prodotto scalare, a coppie di "punti" dello spazio.

### La struttura euclidea: il prodotto scalare

$\mathbb R^n$ diventa uno **spazio euclideo** quando lo dotiamo del prodotto scalare standard (già visto per $\mathbb R^2$ nel notebook 6.2, e generalizzato nel notebook 6.20):

$$x\cdot y = \sum_{i=1}^n x_iy_i = x_1y_1+x_2y_2+\cdots+x_ny_n$$

Questo singolo strumento algebrico ci permetterà, nel prossimo notebook, di definire in modo rigoroso la **norma** (lunghezza) di un vettore e la **distanza** tra due punti — le nozioni che rendono possibile parlare di "vicinanza", indispensabile per estendere il concetto di limite (notebook 5.4) a più variabili.

### Esempi di funzioni di più variabili (anteprima)

Per motivare da subito il resto del modulo, alcuni esempi di funzioni che studieremo:

- $f(x,y)=x^2+y^2$: la distanza al quadrato dall'origine, definita su tutto $\mathbb R^2$.
- $f(x,y)=\sqrt{1-x^2-y^2}$: definita solo dove $x^2+y^2\le1$ (il disco unitario), rappresenta l'emisfero superiore di una sfera.
- $f(x,y,z)=\dfrac1{\sqrt{x^2+y^2+z^2}}$: un modello del potenziale gravitazionale (o elettrico) generato da una massa (o carica) puntiforme nell'origine, definito ovunque tranne che nell'origine stessa.

Formalizzeremo il concetto di **dominio** per queste funzioni (analogo multidimensionale del notebook 3.3) nel notebook 7.4.

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Visualizziamo un punto di R^3 come vettore posizione
punto = np.array([2, 3, 1])
origine = np.array([0, 0, 0])

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')
ax.quiver(*origine, *punto, color='blue', arrow_length_ratio=0.1)
ax.scatter(*punto, color='red', s=50)
ax.set_xlim(0, 3)
ax.set_ylim(0, 3)
ax.set_zlim(0, 3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title("Un punto di R^3 come vettore posizione")
plt.show()

# Prodotto scalare standard in R^3
x = np.array([1, 2, 3])
y = np.array([4, -1, 0])
print("x . y =", np.dot(x, y))

# Esempio: grafico di f(x,y) = x^2 + y^2 (paraboloide)
X = np.linspace(-2, 2, 30)
Y = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(X, Y)
Z = X**2 + Y**2

fig2 = plt.figure(figsize=(6, 6))
ax2 = fig2.add_subplot(111, projection='3d')
ax2.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax2.set_title("f(x,y) = x^2 + y^2")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dato $f(x,y,z)=x^2+y^2-z$, calcola $f$ nei punti $A=(1,1,0)$, $B=(0,0,5)$, $C=(2,-1,3)$, e determina se questi punti soddisfano $f=0$ (cioè se appartengono all'insieme di livello $f(x,y,z)=0$).

**Ragionamento:**

**Passo 1 — calcoliamo $f(A)$**: sostituendo $x=1,y=1,z=0$: $f(A)=1^2+1^2-0=2$. Non è zero: $A$ non appartiene all'insieme di livello.

**Passo 2 — calcoliamo $f(B)$**: $x=0,y=0,z=5$: $f(B)=0+0-5=-5$. Nemmeno $B$ appartiene.

**Passo 3 — calcoliamo $f(C)$**: $x=2,y=-1,z=3$: $f(C)=4+1-3=2$. Anche qui, non zero.

**Conclusione:** nessuno dei tre punti soddisfa $f=0$; interpretando geometricamente, $f(x,y,z)=x^2+y^2-z=0$ (cioè $z=x^2+y^2$) è l'equazione di un **paraboloide** — la superficie che compare nel grafico sopra — e nessuno dei tre punti dati vi giace sopra.

**Perché il calcolo di $f$ in un punto specifico, per funzioni di più variabili, è concettualmente identico al caso di una sola variabile (notebook 3.1), pur avendo più argomenti da sostituire:** valutare $f(x,y,z)$ in un punto richiede semplicemente di sostituire, in ordine, i valori numerici alle variabili — esattamente come per $f(x)$, solo con più sostituzioni simultanee. La novità concettuale non è nel calcolo puntuale, ma in come si comportano **collettivamente** questi valori al variare del punto — dominio, limiti, continuità, derivate — che è l'oggetto dei prossimi notebook.

````python
import sympy as sp

x, y, z = sp.symbols('x y z')
f = x**2 + y**2 - z

punti = {'A': (1, 1, 0), 'B': (0, 0, 5), 'C': (2, -1, 3)}
for nome, (xv, yv, zv) in punti.items():
    valore = f.subs({x: xv, y: yv, z: zv})
    print(f"f({nome}) = f{xv,yv,zv} = {valore}")
````

**1.** $f(1,1)=1-1-1+1=0$. $f(0,2)=0-0-2+1=-1$. $f(2,0)=0-2-0+1=-1$.

## Riepilogo finale

- $\mathbb R^n$ è l'insieme delle $n$-uple di numeri reali, con la doppia natura di insieme di **punti** e di **spazio vettoriale** (notebook 6.3).
- Dotato del prodotto scalare standard $x\cdot y=\sum x_iy_i$ (notebook 6.20), $\mathbb R^n$ diventa uno **spazio euclideo**: la struttura che ci permetterà di misurare lunghezze e distanze.
- Una funzione di più variabili $f(x_1,\dots,x_n)$ si valuta in un punto esattamente come una funzione di una variabile, sostituendo i valori numerici — la complessità nasce dallo studio del comportamento **collettivo**, non dal calcolo puntuale.
- Gli **insiemi di livello** $\{f=c\}$ (curve o superfici) sono uno strumento visivo importante per capire una funzione di più variabili.
