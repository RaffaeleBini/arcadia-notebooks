# 7.12 — Massimi e minimi

**Prerequisiti:** 7.11 Matrice Hessiana, 7.9 Gradiente, 6.24 Teorema spettrale, 5.13 Studio di funzione.


## Perché ci interessa

Nel notebook 5.13, per trovare massimi e minimi di $f(x)$ cercavamo i punti dove $f'(x)=0$, poi usavamo $f''(x)$ per classificarli. In più variabili la strategia è analoga, ma più ricca: annulliamo il **gradiente** (non una sola derivata), e usiamo la **Hessiana** (non un solo numero) per classificare. La novità più interessante è un tipo di punto critico che non esiste in una dimensione: il **punto di sella**, dove la funzione cresce in una direzione e decresce nell'altra — proprio come l'esempio $x^3-3xy^2$ del notebook 7.11 lasciava presagire.

## Teoria, passo dopo passo

### Punti critici

Un punto $x_0$ (interno al dominio) è **critico** per $f$ se $\nabla f(x_0)=\mathbf 0$ (tutte le derivate parziali si annullano). Come nel caso unidimensionale (dove $f'(x_0)=0$ era condizione **necessaria** per un estremo interno, notebook 5.13):

> **Teorema (condizione necessaria).** Se $f$ è differenziabile e ha un massimo o minimo locale in un punto interno $x_0$, allora $\nabla f(x_0)=\mathbf 0$.

**Dimostrazione.** Se $x_0$ è un massimo locale, allora per ogni versore $u$, la funzione di una variabile $g(t)=f(x_0+tu)$ ha un massimo locale in $t=0$; per il notebook 5.13, $g'(0)=0$. Ma $g'(0)=D_uf(x_0)=\nabla f(x_0)\cdot u$ (notebook 7.9). Questo vale per **ogni** $u$: in particolare per $u=e_1,\dots,e_n$, dando $\partial f/\partial x_i(x_0)=0$ per ogni $i$, cioè $\nabla f(x_0)=\mathbf 0$. $\blacksquare$

### Classificazione tramite la Hessiana

Sia $x_0$ un punto critico, con Hessiana $H_f(x_0)$ (simmetrica, notebook 7.11). Per il teorema spettrale (notebook 6.24), $H_f(x_0)$ ha autovalori reali $\lambda_1,\dots,\lambda_n$. Dalla formula di Taylor (notebook 7.11), vicino a $x_0$: $f(x_0+h)-f(x_0)\approx\frac12h^TH_f(x_0)h$ (il termine lineare è nullo, essendo $x_0$ critico). Il segno di questa forma quadratica (notebook 6.23) classifica il punto:

- **definita positiva** (tutti $\lambda_i>0$): $f(x_0+h)>f(x_0)$ per ogni $h$ piccolo $\ne\mathbf 0$ — $x_0$ è un **minimo locale**;
- **definita negativa** (tutti $\lambda_i<0$): $x_0$ è un **massimo locale**;
- **indefinita** (autovalori di segno misto): $f$ cresce in alcune direzioni e decresce in altre — $x_0$ è un **punto di sella**;
- **semidefinita** (qualche $\lambda_i=0$): il test è **inconcludente** — servono informazioni aggiuntive (come visto nell'esempio degenere del notebook 7.11).

### Il test rapido in due variabili

Per $n=2$, usando il test del notebook 6.23 con $D=\det H_f(x_0)=f_{xx}f_{yy}-f_{xy}^2$:

- $D>0$ e $f_{xx}>0$: **minimo locale**;
- $D>0$ e $f_{xx}<0$: **massimo locale**;
- $D<0$: **punto di sella**;
- $D=0$: test **inconcludente**.

### Esempio: il punto di sella

Riprendiamo $f(x,y)=x^2-y^2$ (già incontrata come esempio nel notebook 7.4, con curve di livello iperboliche). $\nabla f=(2x,-2y)=\mathbf 0$ solo in $(0,0)$: unico punto critico. $H_f=\begin{pmatrix}2&0\\0&-2\end{pmatrix}$: $D=2\times(-2)-0=-4<0$: **punto di sella**. Infatti, lungo l'asse $x$ ($y=0$), $f(x,0)=x^2$ cresce allontanandosi dall'origine; lungo l'asse $y$ ($x=0$), $f(0,y)=-y^2$ decresce: comportamento "a sella" (come una sella per cavalli, che sale in una direzione e scende nell'altra).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y', real=True)
f = x**2 - y**2

grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
punti_critici = sp.solve([grad[0], grad[1]], [x, y])
print("Punti critici:", punti_critici)

H = sp.hessian(f, (x, y))
H_critico = H.subs(punti_critici)
print("Hessiana nel punto critico:\n", H_critico)
D = H_critico.det()
print("D = det(H) =", D, "-> punto di sella se D<0")

# Visualizzazione 3D della sella
X = np.linspace(-2, 2, 50)
Y = np.linspace(-2, 2, 50)
X, Y = np.meshgrid(X, Y)
Z = X**2 - Y**2

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='coolwarm', alpha=0.8)
ax.scatter([0], [0], [0], color='black', s=50)
ax.set_title("f(x,y) = x²-y²: punto di sella nell'origine")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova e classifica tutti i punti critici di $f(x,y)=x^3-3x+y^2$.

**Ragionamento:**

**Passo 1 — calcoliamo il gradiente**: $\partial f/\partial x=3x^2-3$, $\partial f/\partial y=2y$.

**Passo 2 — troviamo i punti critici**, risolvendo $\nabla f=\mathbf 0$: da $\partial f/\partial y=2y=0$: $y=0$. Da $\partial f/\partial x=3x^2-3=0$: $x^2=1$, cioè $x=\pm1$ (equazione di secondo grado, notebook 1.6). Due punti critici: $(1,0)$ e $(-1,0)$.

**Passo 3 — calcoliamo la Hessiana generale**: $f_{xx}=6x$, $f_{xy}=0$, $f_{yy}=2$. $H_f(x,y)=\begin{pmatrix}6x&0\\0&2\end{pmatrix}$.

**Passo 4 — classifichiamo $(1,0)$**: $H_f(1,0)=\begin{pmatrix}6&0\\0&2\end{pmatrix}$. $D=6\times2-0=12>0$, $f_{xx}=6>0$: **minimo locale**.

**Passo 5 — classifichiamo $(-1,0)$**: $H_f(-1,0)=\begin{pmatrix}-6&0\\0&2\end{pmatrix}$. $D=-6\times2-0=-12<0$: **punto di sella**.

**Conclusione:** $f$ ha un minimo locale in $(1,0)$ (con $f(1,0)=1-3+0=-2$) e un punto di sella in $(-1,0)$.

**Perché conviene sempre risolvere il sistema $\nabla f=\mathbf 0$ **componente per componente**, isolando prima l'equazione più semplice (qui $\partial f/\partial y=2y=0$, immediata):** quando il sistema è "disaccoppiato" (come qui, dove $\partial f/\partial x$ dipende solo da $x$ e $\partial f/\partial y$ solo da $y$), risolvere prima l'equazione più semplice riduce immediatamente il numero di incognite libere nell'altra equazione — esattamente la stessa strategia di sostituzione già vista per i sistemi lineari (notebook 6.11), applicata qui a un sistema (in generale) non lineare.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)
f = x**3 - 3*x + y**2

grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
punti_critici = sp.solve([grad[0], grad[1]], [x, y])
print("Punti critici:", punti_critici)

H = sp.hessian(f, (x, y))
for pc in punti_critici:
    H_val = H.subs({x: pc[0], y: pc[1]})
    D = H_val.det()
    fxx = H_val[0, 0]
    tipo = "minimo" if (D > 0 and fxx > 0) else "massimo" if (D > 0 and fxx < 0) else "sella" if D < 0 else "inconcludente"
    print(f"Punto {pc}: D={D}, f_xx={fxx} -> {tipo}")
````

## Riepilogo finale

- Un punto critico $x_0$ soddisfa $\nabla f(x_0)=\mathbf 0$: condizione **necessaria** (non sufficiente) per un estremo locale, esattamente come $f'(x_0)=0$ nel caso unidimensionale.
- La **Hessiana** classifica i punti critici tramite il segno della sua forma quadratica (notebook 6.23, 6.24): definita positiva → minimo; definita negativa → massimo; indefinita → **punto di sella** (un fenomeno nuovo, senza analogo in una variabile).
- In due variabili, il test rapido usa $D=\det H_f$ e il segno di $f_{xx}$.
- Quando $D=0$ (o la Hessiana è semidefinita), il test è **inconcludente**: serve un'analisi diretta, come negli esempi degeneri visti.
