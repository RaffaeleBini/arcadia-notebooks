# 7.22 — Rotore

**Prerequisiti:** 7.21 Divergenza, 7.19 Integrali di superficie, 7.20 Campi vettoriali.


## Perché ci interessa

Nel notebook 7.21 la divergenza misurava l'espansione di un campo con un numero. Ma il campo $\mathbf F(x,y)=(-y,x)$ (visto più volte, che "ruota" attorno all'origine) ha divergenza nulla, eppure è tutt'altro che "statico": fa ruotare le cose. Serve un secondo operatore, che catturi la tendenza di un campo a indurre **rotazione** — non un numero stavolta, ma un vettore, che indica l'asse e l'intensità della rotazione locale. È il **rotore**, costruito con lo stesso prodotto vettoriale del notebook 7.19.

## Teoria, passo dopo passo

### Definizione

Per un campo vettoriale $\mathbf F(x,y,z)=(P,Q,R)$ (di classe $C^1$, notebook 7.8), il **rotore** è il campo vettoriale:

$$\operatorname{rot}\mathbf F = \nabla\times\mathbf F = \det\begin{pmatrix}\mathbf i&\mathbf j&\mathbf k\\ \partial/\partial x&\partial/\partial y&\partial/\partial z\\ P&Q&R\end{pmatrix} = \left(\frac{\partial R}{\partial y}-\frac{\partial Q}{\partial z},\ \frac{\partial P}{\partial z}-\frac{\partial R}{\partial x},\ \frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)$$

La notazione ricalca **esattamente** la formula del prodotto vettoriale (notebook 7.19), sostituendo il primo vettore con l'operatore formale $\nabla=(\partial/\partial x,\partial/\partial y,\partial/\partial z)$: uno "sviluppo di Laplace" (notebook 6.12) applicato simbolicamente. Per un campo in $\mathbb R^2$, $\mathbf F=(P,Q)$ (pensato come campo in $\mathbb R^3$ con $R=0$ e nessuna dipendenza da $z$), il rotore ha solo la componente $z$ non banalmente nulla: $\operatorname{rot}\mathbf F=\left(0,0,\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}\right)$ — un vettore perpendicolare al piano, la cui componente $z$ è spesso trattata come uno **scalare** (lo "scalar curl") quando si lavora in due dimensioni.

### Interpretazione fisica: rotazione locale

Se immergiamo una piccola ruota a pale in un fluido con campo di velocità $\mathbf F$, il rotore $\operatorname{rot}\mathbf F(x_0)$ indica **l'asse** attorno a cui la ruota tenderebbe a girare (per la regola della mano destra, coerente con l'orientamento del prodotto vettoriale), e $\|\operatorname{rot}\mathbf F(x_0)\|$ misura **quanto velocemente** girerebbe. Un campo con $\operatorname{rot}\mathbf F=\mathbf 0$ ovunque si dice **irrotazionale**: nessuna tendenza a far ruotare oggetti immersi.

### Campi conservativi sono irrotazionali

Riconosciamo un collegamento notevole col notebook 7.20: se $\mathbf F=\nabla f$ (campo conservativo, con $P=f_x,\,Q=f_y,\,R=f_z$), allora la componente $z$ del rotore è $\partial Q/\partial x-\partial P/\partial y=\partial^2f/\partial x\partial y-\partial^2f/\partial y\partial x=0$ per il teorema di Schwarz (notebook 7.10) — e analogamente per le altre due componenti. Quindi:

> **Ogni campo conservativo è irrotazionale**: $\mathbf F=\nabla f \implies \operatorname{rot}\mathbf F=\mathbf 0$.

Questa è esattamente la stessa condizione necessaria "$\partial P/\partial y=\partial Q/\partial x$" del notebook 7.20, riformulata come annullamento della componente $z$ del rotore — la stessa idea, ora inquadrata nel linguaggio degli operatori differenziali. (Il viceversa vale su domini semplicemente connessi, come vedremo con il teorema di Green.)

### Esempio: il campo rotazionale

Riprendiamo $\mathbf F(x,y)=(-y,x)$: $\operatorname{rot}\mathbf F=\left(0,0,\dfrac{\partial x}{\partial x}-\dfrac{\partial(-y)}{\partial y}\right)=(0,0,1-(-1))=(0,0,2)$. Componente $z$ pari a $2$ (non nulla): coerente con l'osservazione — già fatta nel notebook 7.20 — che questo campo non è conservativo, e ora vediamo esplicitamente **come** ruota (attorno all'asse $z$, in verso antiorario, per il segno positivo).

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)

def rotore(F):
    P, Q, R = F
    return sp.Matrix([
        sp.diff(R, y) - sp.diff(Q, z),
        sp.diff(P, z) - sp.diff(R, x),
        sp.diff(Q, x) - sp.diff(P, y)
    ])

# Campo rotazionale (2D, esteso a R^3 con R=0)
F1 = [-y, x, 0]
print("rot(-y, x, 0) =", rotore(F1).T)

# Campo conservativo: gradiente di f = x^2+y^2+z^2
f = x**2 + y**2 + z**2
F2 = [sp.diff(f, x), sp.diff(f, y), sp.diff(f, z)]
print("F (gradiente di f):", F2)
print("rot(F) =", rotore(F2).T, "-> nullo, coerente: campo conservativo è irrotazionale")
````

## Esercizio svolto e commentato

**Problema:** calcola il rotore di $\mathbf F(x,y,z)=(yz,\,xz,\,xy)$, e verifica se questo campo è conservativo cercandone il potenziale (usando il fatto che il rotore nullo è condizione necessaria).

**Ragionamento:**

**Passo 1 — identifichiamo le componenti**: $P=yz$, $Q=xz$, $R=xy$.

**Passo 2 — calcoliamo la prima componente del rotore**, $\partial R/\partial y-\partial Q/\partial z$: $\partial(xy)/\partial y=x$; $\partial(xz)/\partial z=x$. Prima componente: $x-x=0$.

**Passo 3 — calcoliamo la seconda componente**, $\partial P/\partial z-\partial R/\partial x$: $\partial(yz)/\partial z=y$; $\partial(xy)/\partial x=y$. Seconda componente: $y-y=0$.

**Passo 4 — calcoliamo la terza componente**, $\partial Q/\partial x-\partial P/\partial y$: $\partial(xz)/\partial x=z$; $\partial(yz)/\partial y=z$. Terza componente: $z-z=0$.

**Passo 5 — il rotore è $\mathbf 0$ ovunque**: condizione necessaria per la conservatività soddisfatta. Cerchiamo il potenziale $f$ con $\nabla f=\mathbf F$: integrando $P=yz$ rispetto a $x$: $f=xyz+h(y,z)$.

**Passo 6 — imponiamo $\partial f/\partial y=Q=xz$**: $\partial f/\partial y=xz+\partial h/\partial y$, quindi $\partial h/\partial y=0$, cioè $h=h(z)$ (non dipende da $y$).

**Passo 7 — imponiamo $\partial f/\partial z=R=xy$**: $\partial f/\partial z=xy+h'(z)$, quindi $h'(z)=0$, cioè $h=$ costante.

**Conclusione:** il potenziale è $f(x,y,z)=xyz$ (scegliendo la costante $=0$); verifica: $\nabla f=(yz,xz,xy)=\mathbf F$. ✓ Il campo è conservativo, coerentemente col rotore nullo trovato.

**Perché, con tre variabili, la ricostruzione del potenziale richiede un passo in più rispetto al caso bidimensionale del notebook 7.20 (qui, due "aggiustamenti" successivi, con $h(y,z)$ e poi $h(z)$):** integrando la prima componente rispetto a $x$, la "costante" di integrazione può dipendere da **entrambe** le altre variabili $y,z$ (non solo da una, come nel caso 2D); il procedimento richiede quindi di determinare progressivamente questa funzione arbitraria, prima riducendola a dipendere da una sola variabile (confrontando con $Q$), poi eliminandola del tutto (confrontando con $R$) — un processo che si generalizza naturalmente a un numero qualsiasi di variabili, sempre un passo di "aggiustamento" per ogni componente aggiuntiva del campo.

````python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)
P, Q, R = y*z, x*z, x*y

rot = sp.Matrix([
    sp.diff(R, y) - sp.diff(Q, z),
    sp.diff(P, z) - sp.diff(R, x),
    sp.diff(Q, x) - sp.diff(P, y)
])
print("rot(F) =", rot.T)

f = sp.integrate(P, x)
print("\nPotenziale candidato:", f)
print("Verifica gradiente:", [sp.diff(f, v) for v in (x, y, z)])
````

## Riepilogo finale

- Il **rotore** $\operatorname{rot}\mathbf F=\nabla\times\mathbf F$ è un campo **vettoriale** (a differenza della divergenza, scalare) che misura la tendenza di $\mathbf F$ a indurre rotazione localmente.
- Rotore nullo ovunque $\to$ campo **irrotazionale**; ogni campo conservativo è irrotazionale (conseguenza del teorema di Schwarz, notebook 7.10).
- In due variabili, spesso si usa solo la componente $z$ del rotore (uno scalare), come test rapido di conservatività — coerente con il criterio del notebook 7.20.
- Ricostruire il potenziale in tre variabili richiede due passi di "aggiustamento" successivi, generalizzando il metodo del notebook 7.20.
