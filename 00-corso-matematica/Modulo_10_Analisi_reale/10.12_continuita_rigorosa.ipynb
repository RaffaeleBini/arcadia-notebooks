## Teoria, passo dopo passo

### Continuità: la definizione ε-δ

$$f\text{ è continua in }x_0 \quad\stackrel{\text{def}}{\Longleftrightarrow}\quad \forall\varepsilon>0\,\exists\delta>0:\;|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$$

Confrontala con la definizione di limite (notebook 10.8): è **identica**, tranne per due dettagli. Primo, non serve escludere $x=x_0$ (la condizione $0<|x-x_0|$ diventa $|x-x_0|$, includendo $x_0$), perché in $x_0$ vogliamo proprio confrontare $f(x)$ con $f(x_0)$ stesso, non con un limite esterno. Secondo, il valore di confronto non è un $L$ arbitrario ma **$f(x_0)$**. In altre parole: $f$ è continua in $x_0$ se e solo se $\lim_{x\to x_0}f(x)=f(x_0)$ — la definizione formalizza esattamente ciò che dicevamo informalmente nel notebook 5.6: il limite "prevede" il valore effettivo della funzione.

### Caratterizzazione sequenziale della continuità

**Teorema.** $f$ è continua in $x_0$ $\iff$ per ogni successione $x_n\to x_0$, si ha $f(x_n)\to f(x_0)$.

**Dimostrazione ($\Rightarrow$).** Sia $f$ continua in $x_0$ e $x_n\to x_0$. Fissato $\varepsilon>0$, per continuità esiste $\delta>0$ con $|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$. Poiché $x_n\to x_0$, esiste $N$ con $n>N\Rightarrow|x_n-x_0|<\delta$ (applicando la definizione ε-N del notebook 10.10 con questo $\delta$ come tolleranza). Per $n>N$: $|x_n-x_0|<\delta\Rightarrow|f(x_n)-f(x_0)|<\varepsilon$. Dunque $f(x_n)\to f(x_0)$.

**Dimostrazione ($\Leftarrow$, per contrapposizione, tecnica del notebook 10.6).** Supponiamo $f$ **non** continua in $x_0$: esiste $\varepsilon>0$ tale che per **ogni** $\delta>0$ esiste $x$ con $|x-x_0|<\delta$ ma $|f(x)-f(x_0)|\ge\varepsilon$ (negazione della definizione, notebook 10.2). Scegliendo $\delta=1/n$ per ogni $n\in\mathbb N_{>0}$, otteniamo una successione $x_n$ con $|x_n-x_0|<1/n$ (quindi $x_n\to x_0$, per l'esercizio 1 del notebook 10.10 applicato con $\varepsilon=1/n\to0$) ma $|f(x_n)-f(x_0)|\ge\varepsilon$ per ogni $n$ (quindi $f(x_n)\not\to f(x_0)$). Abbiamo trovato una successione $x_n\to x_0$ con $f(x_n)\not\to f(x_0)$: la condizione a destra è falsa. $\blacksquare$

Questa equivalenza è preziosa: spesso è più comodo lavorare con successioni (specialmente per dimostrare che una funzione **non** è continua) che con la stima diretta ε-δ.

### L'immagine continua di un compatto è compatta

**Teorema.** Se $f$ è continua su $K$ compatto, allora $f(K)$ è compatto.

**Dimostrazione (via compattezza sequenziale, notebook 10.11).** Sia $(y_n)$ una successione in $f(K)$: per ogni $n$, $y_n=f(x_n)$ per qualche $x_n\in K$. Per compattezza sequenziale di $K$, $(x_n)$ ha una sottosuccessione $x_{n_k}\to p\in K$. Per la caratterizzazione sequenziale della continuità (appena dimostrata), $f(x_{n_k})\to f(p)$, cioè $y_{n_k}\to f(p)\in f(K)$. Abbiamo esibito, per una successione arbitraria in $f(K)$, una sottosuccessione convergente a un punto di $f(K)$: per compattezza sequenziale, $f(K)$ è compatto. $\blacksquare$

### Teorema di Weierstrass (dimostrazione completa)

**Teorema.** Se $f:[a,b]\to\mathbb R$ è continua, allora $f$ ammette massimo e minimo su $[a,b]$.

**Dimostrazione.** $[a,b]$ è compatto (chiuso e limitato, Heine-Borel, notebook 10.11). Per il teorema appena dimostrato, $f([a,b])$ è compatto, quindi chiuso e limitato. Essendo limitato e non vuoto, $M=\sup f([a,b])$ esiste (assioma di completezza, notebook 10.9). Poiché $f([a,b])$ è **chiuso**, contiene tutti i limiti delle proprie successioni convergenti (definizione di chiuso, notebook 7.3); per la condizione 2 della definizione di $\sup$ (notebook 10.9), esiste una successione $y_n\in f([a,b])$ con $y_n\to M$ — quindi $M$, come limite di una successione in $f([a,b])$, appartiene a $f([a,b])$ perché l'insieme è chiuso. Dunque esiste $x^*\in[a,b]$ con $f(x^*)=M$: $M$ è **raggiunto**, cioè è il massimo (non solo l'estremo superiore). Lo stesso argomento con $\inf$ dà il minimo. $\blacksquare$

**Il punto cruciale**, ora visibile con chiarezza: nel notebook 10.9 avevamo osservato che $\sup A$ non deve appartenere ad $A$ (esempio $A=\{1-1/n\}$, $\sup A=1\notin A$). Qui, invece, la **chiusura** di $f([a,b])$ (garantita dalla compattezza del dominio) forza $M=\sup f([a,b])$ a essere raggiunto. È esattamente questo il ruolo della compattezza: trasforma un estremo superiore "potenzialmente irraggiungibile" in un vero massimo.

````python
import numpy as np
import matplotlib.pyplot as plt

# Illustrazione: f(x) = x^3 - 3x su [-2, 2] (compatto), continua, ammette massimo e minimo
x = np.linspace(-2, 2, 400)
f = x**3 - 3*x

x_max_idx = np.argmax(f)
x_min_idx = np.argmin(f)

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(x, f, color="steelblue")
ax.plot(x[x_max_idx], f[x_max_idx], "o", color="green", label=f"massimo raggiunto in x={x[x_max_idx]:.2f}")
ax.plot(x[x_min_idx], f[x_min_idx], "o", color="red", label=f"minimo raggiunto in x={x[x_min_idx]:.2f}")
ax.set_title("f(x) = x^3 - 3x su [-2,2]: dominio compatto, massimo e minimo esistono e sono raggiunti")
ax.legend()
ax.axhline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()

print("Massimo:", f[x_max_idx], "raggiunto in x =", x[x_max_idx])
print("Minimo:", f[x_min_idx], "raggiunto in x =", x[x_min_idx])
````

## Esercizio svolto e commentato

**Problema:** dimostra, usando la caratterizzazione sequenziale, che $f(x)=x^2$ è continua in ogni $x_0\in\mathbb R$.

**Ragionamento:**

**Passo 1 — fissiamo $x_0$ arbitrario e una successione $x_n\to x_0$.** Dobbiamo dimostrare $f(x_n)=x_n^2\to f(x_0)=x_0^2$.

**Passo 2 — usiamo l'algebra dei limiti (notebook 10.10).** Sappiamo che se $a_n\to L$ e $b_n\to M$, allora $a_nb_n\to LM$ (regola del prodotto, dimostrabile con lo stesso schema "dividi $\varepsilon$" della regola della somma, con una stima leggermente più elaborata che usa la limitatezza delle successioni convergenti — esercizio 3 del notebook 10.10).

**Passo 3 — applichiamo la regola con $a_n=b_n=x_n$.** Poiché $x_n\to x_0$ per ipotesi, la regola del prodotto dà $x_n\cdot x_n\to x_0\cdot x_0$, cioè $x_n^2\to x_0^2$.

**Conclusione:** $f(x_n)\to f(x_0)$ per ogni successione $x_n\to x_0$: per la caratterizzazione sequenziale, $f(x)=x^2$ è continua in $x_0$. Essendo $x_0$ arbitrario, $f$ è continua su tutto $\mathbb R$. **Perché questo approccio è più rapido della stima ε-δ diretta** (fatta nel notebook 10.8 per $x_0=3$, con il "trucco del minimo"): la caratterizzazione sequenziale permette di riciclare l'algebra dei limiti già dimostrata, invece di rifare da capo una stima con fattori da limitare — un vantaggio che cresce man mano che le funzioni diventano più complesse (somme, prodotti, composizioni di funzioni già note continue).

````python
import numpy as np

def f(x):
    return x**2

x0 = 3.0
# Successione arbitraria che tende a x0
n = np.arange(1, 100)
xn = x0 + 1.0/n  # tende a x0 da destra

print("x_n (primi termini):", xn[:5], "-> tende a", x0)
print("f(x_n) (primi termini):", f(xn[:5]))
print("f(x0) =", f(x0))
print("f(x_n) tende a f(x0)?", np.isclose(f(xn[-1]), f(x0), atol=1e-3))
````

**1.** Sia $x_n\to x_0$. Per continuità di $f$ e $g$ (caratterizzazione sequenziale), $f(x_n)\to f(x_0)$ e $g(x_n)\to g(x_0)$. Per la regola della somma (notebook 10.10), $f(x_n)+g(x_n)\to f(x_0)+g(x_0)$, cioè $(f+g)(x_n)\to(f+g)(x_0)$. Essendo $x_n\to x_0$ arbitraria, per la caratterizzazione sequenziale $f+g$ è continua in $x_0$.

**2.** Sia $x_n=-1/n\to0$. Allora $f(x_n)=\operatorname{sgn}(-1/n)=-1$ per ogni $n$, quindi $f(x_n)\to-1\ne0=f(0)$. Trovata una successione $x_n\to0$ con $f(x_n)\not\to f(0)$: per la caratterizzazione sequenziale, $f$ non è continua in $0$.

**3.** Per il teorema della teoria, $f(K)$ è compatto (immagine continua di un compatto), quindi per Heine-Borel (notebook 10.11) $f(K)$ è limitato: esiste $M$ con $|f(x)|\le M$ per ogni $x\in K$, cioè $f$ è limitata su $K$.

**4.** Sia $A=\{x\in[a,b]:f(x)<0\}$: non vuoto ($a\in A$, perché $f(a)<0$) e limitato superiormente (da $b$), quindi $c=\sup A$ esiste per completezza (notebook 10.9); inoltre $c\in[a,b]$ (perché $a\in A\Rightarrow c\ge a$, e $b$ è un maggiorante di $A\Rightarrow c\le b$). **Se $f(c)<0$**: per continuità, esiste $\delta>0$ tale che $f$ resta negativa su $(c-\delta,c+\delta)\cap[a,b]$ (un fatto che segue dalla definizione ε-δ, scegliendo $\varepsilon=|f(c)|$); in particolare $c<b$ (altrimenti $f(b)<0$, contro l'ipotesi $f(b)>0$), quindi esistono punti di $A$ oltre $c$, contraddicendo che $c$ fosse un maggiorante. **Se $f(c)>0$**: analogamente, per continuità $f$ resta positiva su un intorno di $c$, quindi $c-\delta$ è un maggiorante di $A$ più piccolo di $c$ (nessun punto di $A$ può stare in $(c-\delta,c]$), contraddicendo la minimalità di $c=\sup A$. Restano escluse entrambe le alternative: $f(c)=0$, con $c\in(a,b)$ (essendo $f(a)<0$ e $f(b)>0$, non può essere $c=a$ né $c=b$).

## Riepilogo finale

- La definizione ε-δ di continuità dice $f(x)\to f(x_0)$ per $x\to x_0$: identica al limite, ma il valore di confronto è $f(x_0)$ stesso.
- La **caratterizzazione sequenziale** ($x_n\to x_0\Rightarrow f(x_n)\to f(x_0)$) è equivalente ed è spesso più comoda, specialmente per dimostrare che una funzione **non** è continua.
- L'**immagine continua di un compatto è compatta**: il teorema chiave che collega continuità e compattezza.
- Il **teorema di Weierstrass** segue immediatamente: $f([a,b])$ compatto, quindi chiuso e limitato, quindi il suo $\sup$ vi appartiene — è un vero massimo, raggiunto.
