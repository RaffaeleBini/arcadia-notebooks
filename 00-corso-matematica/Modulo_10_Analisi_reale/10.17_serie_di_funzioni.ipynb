## Teoria, passo dopo passo

### Serie di funzioni: convergenza puntuale e uniforme

Data una successione di funzioni $g_n:E\to\mathbb R$, la **serie di funzioni** $\sum_{n=0}^\infty g_n(x)$ è, per definizione, il limite delle somme parziali $S_N(x)=\sum_{n=0}^Ng_n(x)$ — esattamente come per le serie numeriche del notebook 4.7, ma ora $x$ è una variabile. La serie converge **uniformemente** su $E$ se la successione di funzioni $(S_N)$ converge uniformemente (notebook 10.16) alla funzione somma $S(x)=\sum_ng_n(x)$.

### Il criterio di Weierstrass (M-test)

**Teorema.** Se esiste una successione di costanti $M_n\ge0$ tali che $|g_n(x)|\le M_n$ per **ogni** $x\in E$, e $\sum_nM_n$ converge, allora $\sum_ng_n(x)$ converge **uniformemente** (e assolutamente) su $E$.

**Dimostrazione.** Per ogni $x\in E$, $|g_n(x)|\le M_n$ e $\sum M_n$ converge, quindi per il criterio del confronto (notebook 4.10) $\sum_n|g_n(x)|$ converge — cioè la serie converge assolutamente in ogni punto, quindi converge puntualmente a una funzione $S(x)$. Per l'uniformità, stimiamo il resto: $|S(x)-S_N(x)|=\left|\sum_{n=N+1}^\infty g_n(x)\right|\le\sum_{n=N+1}^\infty|g_n(x)|\le\sum_{n=N+1}^\infty M_n$. Quest'ultima è la coda di una serie numerica **convergente**, quindi tende a $0$ per $N\to\infty$ (definizione di convergenza di una serie, notebook 4.7) — **indipendentemente da $x$**: dato $\varepsilon>0$, esiste $N$ (che dipende solo da $\varepsilon$, tramite la coda della serie di costanti) tale che $\sum_{n=N+1}^\infty M_n<\varepsilon$, quindi $|S(x)-S_N(x)|<\varepsilon$ per ogni $x\in E$ simultaneamente: convergenza uniforme. $\blacksquare$

**Perché questo criterio è così efficace**: riduce un problema sulla variabile $x$ (potenzialmente complicato) a un problema puramente numerico su $\sum M_n$ — spesso una $p$-serie o una serie geometrica, già ben comprese dal notebook 4.8-4.10.

### Derivazione e integrazione termine a termine: ora giustificate

**Teorema (integrazione termine a termine).** Se $\sum g_n$ converge uniformemente su $[a,b]$ e ogni $g_n$ è integrabile, allora $\int_a^b\sum_ng_n(x)\,dx=\sum_n\int_a^bg_n(x)\,dx$.

Segue immediatamente applicando alle somme parziali $S_N$ il teorema di scambio limite-integrale dimostrato nel notebook 10.16 (esercizio 4): $\int_a^bS_N\to\int_a^bS$, cioè $\sum_{n=0}^N\int_a^bg_n\to\int_a^bS$, che è precisamente l'uguaglianza cercata, passando al limite $N\to\infty$.

**Teorema (derivazione termine a termine, enunciato).** Se $\sum g_n(x_0)$ converge in almeno un punto, ogni $g_n$ è derivabile con derivata continua, e $\sum g_n'$ converge **uniformemente**, allora $\sum g_n$ converge (uniformemente) a una funzione derivabile, e $\left(\sum_ng_n\right)'=\sum_ng_n'$.

Nota bene: a differenza dell'integrazione, per la derivazione l'ipotesi cruciale è sulla convergenza uniforme **delle derivate**, non della serie originale — esattamente la lezione dell'esempio $\sin(nx)/n$ del notebook 10.16, dove la convergenza uniforme della funzione non impediva il collasso delle derivate.

### Serie di potenze: convergenza uniforme sui compatti interni

Una serie di potenze $\sum a_nx^n$ ha un raggio di convergenza $R$ (notebook 4.11): converge assolutamente per $|x|<R$. Applicando Weierstrass su un qualsiasi $[-r,r]$ con $r<R$: poiché $|a_nx^n|\le|a_n|r^n$ per $|x|\le r$, e $\sum|a_n|r^n$ converge (essendo $r<R$, dentro il raggio di convergenza), il criterio di Weierstrass garantisce convergenza **uniforme** su $[-r,r]$. Questo giustifica finalmente, con rigore, la derivazione e integrazione termine a termine delle serie di Taylor usate liberamente dal notebook 4.12 in poi — incluse le dimostrazioni della formula di Eulero e delle proprietà di $e^x,\sin x,\cos x$ nel notebook 9.3, ora pienamente fondate.

### Serie di Fourier: perché la convergenza uniforme può fallire

Le somme parziali di Fourier (notebook 8.9) di un segnale con **discontinuità a salto** (come l'onda quadra) **non** convergono uniformemente vicino al salto — un fenomeno noto come **fenomeno di Gibbs**: le somme parziali sviluppano un'oscillazione ("orecchio") di ampiezza quasi costante (circa il 9% del salto) che non si attenua aumentando i termini, per quanto si restringa in larghezza. Questo è del tutto coerente con il teorema del notebook 10.16: se la convergenza fosse uniforme, il limite (discontinuo) dovrebbe essere continuo — contraddizione. La convergenza delle serie di Fourier con salti è solo **puntuale** (anzi, in un senso più debole, "quasi ovunque"), mai uniforme vicino alle discontinuità — un limite intrinseco che nessun perfezionamento del calcolo può aggirare, e che ora, con gli strumenti di questo modulo, sappiamo spiegare invece di limitarci a osservare.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-np.pi, np.pi, 2000)

def onda_quadra_parziale(x, N):
    somma = np.zeros_like(x)
    for k in range(1, N + 1, 2):  # solo armoniche dispari
        somma += (4 / (np.pi * k)) * np.sin(k * x)
    return somma

fig, ax = plt.subplots(figsize=(8, 4.5))
for N in [5, 21, 101]:
    ax.plot(x, onda_quadra_parziale(x, N), label=f"N={N} termini")

ax.axhline(1, color="gray", linestyle="--", linewidth=0.7)
ax.axhline(-1, color="gray", linestyle="--", linewidth=0.7)
ax.set_title("Fenomeno di Gibbs: l'oscillazione vicino al salto NON si attenua con N")
ax.legend()
plt.tight_layout()
plt.show()

# Verifica numerica: l'overshoot massimo resta circa costante (~9%) al crescere di N
for N in [11, 51, 201]:
    somma = onda_quadra_parziale(x, N)
    overshoot = (somma.max() - 1) / 1 * 100
    print(f"N={N}: overshoot massimo vicino al salto = {overshoot:.2f}% (atteso: resta vicino a 9%, non diminuisce)")
````

## Esercizio svolto e commentato

**Problema:** usa il criterio di Weierstrass per dimostrare che $\sum_{n=1}^\infty\dfrac{\cos(nx)}{n^2}$ converge uniformemente su tutto $\mathbb R$, e che la funzione somma è continua ovunque.

**Ragionamento:**

**Passo 1 — troviamo una maggiorazione indipendente da $x$.** Per ogni $x\in\mathbb R$ e ogni $n$: $\left|\dfrac{\cos(nx)}{n^2}\right|=\dfrac{|\cos(nx)|}{n^2}\le\dfrac1{n^2}$ (usando $|\cos(nx)|\le1$, notebook 2.9). Poniamo $M_n=\dfrac1{n^2}$.

**Passo 2 — verifichiamo che $\sum M_n$ converge.** $\sum_{n=1}^\infty\dfrac1{n^2}$ è una $p$-serie con $p=2>1$, convergente (criterio delle $p$-serie, notebook 4.10) — anzi, converge esattamente a $\pi^2/6$, come già visto nel notebook 10.10.

**Passo 3 — applichiamo il criterio di Weierstrass.** Le ipotesi sono soddisfatte ($|g_n(x)|\le M_n$ per ogni $x$, $\sum M_n$ convergente): la serie $\sum\dfrac{\cos(nx)}{n^2}$ converge **uniformemente** su tutto $\mathbb R$.

**Passo 4 — deduciamo la continuità della somma.** Ogni termine $g_n(x)=\dfrac{\cos(nx)}{n^2}$ è continuo (composizione di funzioni continue). Per il teorema del notebook 10.16 (limite uniforme di funzioni continue è continuo), applicato alla successione delle somme parziali $S_N$ (ciascuna continua, come somma finita di funzioni continue), la funzione somma $S(x)=\sum_{n=1}^\infty\dfrac{\cos(nx)}{n^2}$ è **continua su tutto $\mathbb R$**.

**Conclusione:** questa serie — che è, in effetti, una serie di Fourier — rappresenta una funzione **continua** (a differenza dell'onda quadra della teoria), coerentemente con il fatto che i suoi coefficienti decadono abbastanza rapidamente ($1/n^2$, sommabile) da garantire la convergenza uniforme, mentre i coefficienti dell'onda quadra ($1/n$, non sommabile in valore assoluto) non ci riescono — la differenza tra $1/n^2$ e $1/n$, apparentemente un dettaglio tecnico, è precisamente ciò che separa una somma continua da una con salti.

````python
import numpy as np

def somma_parziale(x, N):
    n = np.arange(1, N + 1)
    return np.sum(np.cos(n[:, None] * x) / n[:, None]**2, axis=0)

x = np.linspace(-2*np.pi, 2*np.pi, 500)

# Verifica che le somme parziali si stabilizzino rapidamente (convergenza uniforme -> veloce)
for N in [5, 20, 100]:
    S_N = somma_parziale(x, N)
    S_grande = somma_parziale(x, 500)  # approssimazione della somma "vera"
    scarto_massimo = np.max(np.abs(S_N - S_grande))
    print(f"N={N}: scarto massimo rispetto alla somma con 500 termini = {scarto_massimo:.6f}")

print("\nLo scarto massimo (su TUTTO il dominio) decresce con N: è proprio la convergenza uniforme.")
print("Confronta con il fenomeno di Gibbs sopra, dove l'overshoot NON decresce.")
````

Il membro sinistro, calcolato direttamente, è $[-\ln(1-x)]_0^r=-\ln(1-r)$. Dunque $-\ln(1-r)=\sum_{n=1}^\infty\dfrac{r^n}n$.

## Riepilogo finale

- Il **criterio di Weierstrass** (M-test) riduce la convergenza uniforme di una serie di funzioni a quella di una serie numerica di maggioranti $\sum M_n$: uno strumento potente e semplice da applicare.
- Sotto convergenza uniforme, **l'integrazione termine a termine è sempre lecita**; la **derivazione termine a termine** richiede l'ipotesi più forte che siano le **derivate** a convergere uniformemente.
- Le serie di potenze convergono uniformemente su ogni compatto interno al raggio di convergenza, fondando rigorosamente tutto il calcolo con le serie di Taylor del Modulo 4 e del Modulo 9.
- Il **fenomeno di Gibbs** nelle serie di Fourier di funzioni discontinue è la manifestazione concreta di un fatto ora pienamente compreso: la convergenza non può essere uniforme vicino a un salto, perché il limite uniforme di funzioni continue è sempre continuo.
