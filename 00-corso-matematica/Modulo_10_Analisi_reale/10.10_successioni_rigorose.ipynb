## Teoria, passo dopo passo

### La definizione ε-N

$$\lim_{n\to\infty}a_n=L \quad\stackrel{\text{def}}{\Longleftrightarrow}\quad \forall\varepsilon>0\,\exists N\in\mathbb N:\; n>N\Rightarrow|a_n-L|<\varepsilon$$

Stessa struttura logica del notebook 10.8, con $N$ al posto di $\delta$: dato un margine di tolleranza $\varepsilon$, da un certo indice $N$ in poi (che può dipendere da $\varepsilon$) tutti i termini della successione restano vicini a $L$ meno di $\varepsilon$.

### Unicità del limite

**Teorema.** Se $a_n\to L_1$ e $a_n\to L_2$, allora $L_1=L_2$.

**Dimostrazione (per assurdo).** Supponiamo $L_1\ne L_2$, e poniamo $\varepsilon=\dfrac{|L_1-L_2|}2>0$. Per la definizione applicata a $L_1$, esiste $N_1$ con $n>N_1\Rightarrow|a_n-L_1|<\varepsilon$. Per la definizione applicata a $L_2$, esiste $N_2$ con $n>N_2\Rightarrow|a_n-L_2|<\varepsilon$. Per $n>\max(N_1,N_2)$, per la disuguaglianza triangolare (notebook 0.8): $|L_1-L_2|\le|L_1-a_n|+|a_n-L_2|<\varepsilon+\varepsilon=2\varepsilon=|L_1-L_2|$. Otteniamo $|L_1-L_2|<|L_1-L_2|$, assurdo. Dunque $L_1=L_2$. $\blacksquare$

### La regola della somma (esempio di algebra dei limiti)

**Teorema.** Se $a_n\to L$ e $b_n\to M$, allora $a_n+b_n\to L+M$.

**Dimostrazione.** Fissato $\varepsilon>0$, applichiamo la definizione ad $a_n$ con tolleranza $\varepsilon/2$: esiste $N_1$ con $n>N_1\Rightarrow|a_n-L|<\varepsilon/2$. Analogamente per $b_n$: esiste $N_2$ con $n>N_2\Rightarrow|b_n-M|<\varepsilon/2$. Per $n>N=\max(N_1,N_2)$: $|(a_n+b_n)-(L+M)|=|(a_n-L)+(b_n-M)|\le|a_n-L|+|b_n-M|<\dfrac\varepsilon2+\dfrac\varepsilon2=\varepsilon$ (disuguaglianza triangolare). $\blacksquare$ Le altre regole dell'algebra dei limiti (prodotto, quoziente) si dimostrano con lo stesso schema — "dividere $\varepsilon$" tra i pezzi — con stime leggermente più elaborate per il prodotto.

### Teorema di monotonia (dimostrazione completa)

**Teorema.** Ogni successione $(a_n)$ monotona crescente e limitata superiormente converge, e $\lim a_n=\sup\{a_n:n\in\mathbb N\}$.

**Dimostrazione.** Sia $A=\{a_n:n\in\mathbb N\}$: non vuoto e limitato superiormente per ipotesi, quindi per l'assioma di completezza (notebook 10.9) esiste $s=\sup A$. Dimostriamo $a_n\to s$. Fissato $\varepsilon>0$: per la condizione 2 della definizione di $\sup$ (notebook 10.9), esiste $N$ con $a_N>s-\varepsilon$. Per monotonia, $n>N\Rightarrow a_n\ge a_N>s-\varepsilon$. Inoltre $a_n\le s$ per ogni $n$ (perché $s$ è un maggiorante), quindi $a_n<s+\varepsilon$ banalmente. Combinando: $s-\varepsilon<a_n\le s<s+\varepsilon$, cioè $|a_n-s|<\varepsilon$ per ogni $n>N$. $\blacksquare$

### Teorema di Bolzano-Weierstrass (dimostrazione via bisezione)

**Teorema.** Ogni successione limitata $(a_n)$ ammette una sottosuccessione convergente.

**Dimostrazione.** Sia $(a_n)\subseteq[c,d]$ (limitata, per ipotesi). Costruiamo per bisezione una successione di intervalli incapsulati: $I_0=[c,d]$. Dividiamo $I_0$ a metà: almeno una delle due metà contiene $a_n$ per **infiniti** indici $n$ (se entrambe ne contenessero solo un numero finito, l'intera successione avrebbe solo un numero finito di termini — assurdo, dato che $(a_n)$ ha infiniti termini). Scegliamo $I_1$ una tale metà, e scegliamo $n_1$ con $a_{n_1}\in I_1$. Ripetiamo: dividiamo $I_1$ a metà, scegliamo $I_2$ la metà con infiniti indici, e $n_2>n_1$ con $a_{n_2}\in I_2$. Iterando, otteniamo intervalli incapsulati $I_0\supseteq I_1\supseteq I_2\supseteq\cdots$, ciascuno di lunghezza $\dfrac{d-c}{2^k}$, e indici crescenti $n_1<n_2<n_3<\cdots$ con $a_{n_k}\in I_k$.

Per il **teorema degli intervalli incapsulati** (notebook 10.9), esiste $L\in\bigcap_kI_k$. Dimostriamo $a_{n_k}\to L$: sia $\varepsilon>0$; poiché la lunghezza di $I_k$ tende a $0$ (essendo $\dfrac{d-c}{2^k}\to0$, un fatto che segue dalla proprietà archimedea), esiste $K$ con $\dfrac{d-c}{2^K}<\varepsilon$; per $k>K$, sia $a_{n_k}$ sia $L$ appartengono a $I_k$ (per $L$: perché $L\in I_j$ per ogni $j$; per $a_{n_k}$: per costruzione), quindi $|a_{n_k}-L|\le$ lunghezza di $I_k<\varepsilon$. Dunque $(a_{n_k})_k$ è la sottosuccessione convergente cercata. $\blacksquare$

### Successioni di Cauchy e il criterio di Cauchy

Una successione $(a_n)$ è di **Cauchy** se $\forall\varepsilon>0\,\exists N:\;n,m>N\Rightarrow|a_n-a_m|<\varepsilon$ — i termini si avvicinano **tra loro**, senza fare riferimento a un limite noto a priori.

**Teorema (criterio di Cauchy).** $(a_n)$ converge $\iff$ $(a_n)$ è di Cauchy.

**Dimostrazione ($\Rightarrow$).** Se $a_n\to L$: fissato $\varepsilon>0$, esiste $N$ con $n>N\Rightarrow|a_n-L|<\varepsilon/2$. Per $n,m>N$: $|a_n-a_m|\le|a_n-L|+|L-a_m|<\varepsilon/2+\varepsilon/2=\varepsilon$ (disuguaglianza triangolare).

**Dimostrazione ($\Leftarrow$), in due passi.** *Passo 1: ogni successione di Cauchy è limitata* (con $\varepsilon=1$, da un certo $N$ in poi i termini distano meno di $1$ tra loro, quindi sono confinati in un intervallo finito; i primi $N$ termini, essendo in numero finito, sono anch'essi limitati). *Passo 2*: essendo limitata, per **Bolzano-Weierstrass** (appena dimostrato) $(a_n)$ ha una sottosuccessione $a_{n_k}\to L$. Dimostriamo che l'**intera** successione converge a $L$: fissato $\varepsilon>0$, per la proprietà di Cauchy esiste $N_1$ con $n,m>N_1\Rightarrow|a_n-a_m|<\varepsilon/2$; per la convergenza della sottosuccessione esiste $k$ con $n_k>N_1$ e $|a_{n_k}-L|<\varepsilon/2$; allora per ogni $n>N_1$: $|a_n-L|\le|a_n-a_{n_k}|+|a_{n_k}-L|<\varepsilon/2+\varepsilon/2=\varepsilon$ (usando $n,n_k>N_1$ nella proprietà di Cauchy). Dunque $a_n\to L$. $\blacksquare$

**Perché questo criterio è così importante:** permette di dimostrare che una successione converge **senza conoscere il limite** — un vantaggio enorme quando il limite non ha un'espressione chiusa. Il criterio di Cauchy è, in realtà, un'altra formulazione equivalente della completezza di $\mathbb R$: uno spazio in cui ogni successione di Cauchy converge si dice **completo**, e $\mathbb Q$ non lo è (una successione di razionali che approssima $\sqrt2$ sempre meglio è di Cauchy in $\mathbb Q$, ma non converge a nessun razionale).

````python
import numpy as np

# Verifica numerica di Bolzano-Weierstrass su una successione limitata ma non convergente: a_n = sin(n)
n = np.arange(1, 5000)
a = np.sin(n)

print("La successione sin(n) e' limitata:", np.all(np.abs(a) <= 1))

# Estraiamo una sottosuccessione convergente cercando indici dove sin(n) e' vicino a 1
vicini_a_1 = n[np.abs(a - 1) < 0.001]
print("\nPrimi indici n con sin(n) vicino a 1:", vicini_a_1[:5])
print("Valori corrispondenti:", np.sin(vicini_a_1[:5]))
print("Bolzano-Weierstrass garantisce che una tale sottosuccessione (convergente a 1) esiste sempre,")
print("anche se l'intera successione sin(n) non converge.")

# Verifica del criterio di Cauchy: a_n = somma parziale di 1/2^k (converge, quindi e' di Cauchy)
somme_parziali = np.cumsum([1/2**k for k in range(1, 50)])
print("\nDifferenze |a_n - a_m| per n,m grandi (attese piccole, successione di Cauchy):")
print(abs(somme_parziali[-1] - somme_parziali[-5]))
````

## Esercizio svolto e commentato

**Problema:** dimostra, usando il criterio di Cauchy, che la successione $a_n=\displaystyle\sum_{k=1}^n\dfrac1{k^2}$ converge (senza calcolarne il limite esatto).

**Ragionamento:**

**Passo 1 — impostiamo la stima di Cauchy.** Per $m>n$: $|a_m-a_n|=\displaystyle\sum_{k=n+1}^m\dfrac1{k^2}$.

**Passo 2 — maggioriamo ogni termine con qualcosa di telescopico.** Per $k\ge2$, $\dfrac1{k^2}<\dfrac1{k(k-1)}=\dfrac1{k-1}-\dfrac1k$ (scomposizione in fratti semplici, verificabile per somma di frazioni, notebook 1.4). Quindi, per $n\ge1$:

$$\sum_{k=n+1}^m\frac1{k^2}<\sum_{k=n+1}^m\left(\frac1{k-1}-\frac1k\right)$$

**Passo 3 — la somma telescopica si semplifica.** La somma a destra è **telescopica** (ogni termine cancella parte del successivo, un meccanismo già visto nel notebook 4.7 per le serie numeriche): $\displaystyle\sum_{k=n+1}^m\left(\frac1{k-1}-\frac1k\right)=\frac1n-\frac1m<\frac1n$.

**Passo 4 — concludiamo la stima di Cauchy.** Per ogni $m>n\ge1$: $|a_m-a_n|<\dfrac1n$. Dato $\varepsilon>0$, per la proprietà archimedea (notebook 10.9) esiste $N$ con $\dfrac1N<\varepsilon$; per $n,m>N$ (assumendo $m>n$ senza perdita di generalità, essendo $|a_m-a_n|$ simmetrico): $|a_m-a_n|<\dfrac1n<\dfrac1N<\varepsilon$.

**Conclusione:** $(a_n)$ è di Cauchy, quindi per il criterio di Cauchy converge. **Il valore aggiunto del criterio:** abbiamo dimostrato la convergenza senza mai calcolare $\lim a_n$ — che in effetti vale $\pi^2/6$ (il celebre problema di Basilea), un fatto che richiede strumenti ben più elaborati delle sole stime di Cauchy.

````python
import numpy as np

def somma_parziale(n):
    return sum(1.0/k**2 for k in range(1, n+1))

# Verifica della stima |a_m - a_n| < 1/n per alcune coppie
for n, m in [(10, 20), (100, 500), (1000, 10000)]:
    diff = abs(somma_parziale(m) - somma_parziale(n))
    print(f"n={n}, m={m}: |a_m - a_n| = {diff:.6f}, stima 1/n = {1/n:.6f}, "
          f"stima rispettata? {diff < 1/n}")

print("\nValore a cui converge (Problema di Basilea):", np.pi**2/6)
print("Somma parziale fino a n=100000:", somma_parziale(100000))
````

**4.** Per $n\ge2$: $a_{n+1}-a_n=\dfrac1{a_n}-\dfrac1{a_{n-1}}=\dfrac{a_{n-1}-a_n}{a_na_{n-1}}$ (differenza di frazioni, notebook 1.4). ## Riepilogo finale

- La definizione ε-N di limite di successione ha la stessa struttura logica del limite ε-δ di funzione (notebook 10.8); da essa si dimostrano rigorosamente **unicità del limite** e **algebra dei limiti**.
- Il **teorema di monotonia** (successione monotona e limitata converge) si dimostra direttamente dall'assioma di completezza: il limite è semplicemente il $\sup$ (o $\inf$).
- Il **teorema di Bolzano-Weierstrass** (ogni successione limitata ha una sottosuccessione convergente) si dimostra per **bisezione**, usando il teorema degli intervalli incapsulati del notebook 10.9.
- Il **criterio di Cauchy** caratterizza la convergenza senza conoscere il limite in anticipo, ed è esso stesso equivalente alla completezza di $\mathbb R$: $\mathbb Q$ non è completo in questo senso.
