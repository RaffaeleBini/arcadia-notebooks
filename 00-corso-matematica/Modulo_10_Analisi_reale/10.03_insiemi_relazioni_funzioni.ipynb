## Perché ci interessa

Nel notebook 0.12 abbiamo definito un insieme come "una collezione di oggetti", e nel notebook 3.1 una funzione come "una regola che associa a ogni elemento di un insieme uno e un solo elemento di un altro". Sono descrizioni intuitive, sufficienti per fare calcoli, ma non abbastanza precise per fondare dimostrazioni rigorose: cosa significa esattamente "regola"? E cosa vuol dire che due insiemi "hanno lo stesso numero di elementi" quando sono infiniti, come $\mathbb N$ e $\mathbb Q$ (domanda a cui risponderemo nel notebook 10.5)? Per rispondere servono fondamenta più solide: la nozione di **relazione** tra insiemi, di cui la funzione è un caso particolare, e le relazioni di **equivalenza** e **d'ordine**, che formalizzano idee come "avere la stessa dimensione" o "essere confrontabili" già usate implicitamente in tutto il corso.

## Teoria, passo dopo passo

### Coppie ordinate e prodotto cartesiano

Il **prodotto cartesiano** $A\times B$ di due insiemi è l'insieme di tutte le coppie ordinate $(a,b)$ con $a\in A$, $b\in B$ — la stessa costruzione che dà origine al piano cartesiano $\mathbb R\times\mathbb R$ (notebook 2.1). "Ordinata" significa che $(a,b)\ne(b,a)$ in generale: l'ordine conta, a differenza degli insiemi $\{a,b\}=\{b,a\}$ (notebook 0.12).

### Relazione: la nozione più generale

Una **relazione** $R$ tra $A$ e $B$ è semplicemente un sottoinsieme $R\subseteq A\times B$. Scriviamo $a\,R\,b$ per dire $(a,b)\in R$. Questa definizione, puramente insiemistica, cattura ogni legame possibile tra elementi di $A$ e di $B$: "essere minore di", "essere multiplo di", "essere il quadrato di" sono tutte relazioni.

### Funzione come caso particolare di relazione

Una **funzione** $f:A\to B$ è una relazione $R\subseteq A\times B$ che soddisfa due condizioni:

1. **totalità**: per ogni $a\in A$ esiste almeno un $b\in B$ con $a\,R\,b$ (ogni elemento del dominio ha un'immagine);
2. **unicità**: se $a\,R\,b_1$ e $a\,R\,b_2$, allora $b_1=b_2$ (l'immagine è unica).

Questa è la definizione rigorosa che sostituisce la "regola" informale del notebook 3.1: una funzione **è** l'insieme delle sue coppie $(x,f(x))$ — esattamente il suo grafico, già disegnato in tanti notebook precedenti, ora elevato a definizione. Iniettività e suriettività (notebook 3.5) si riformulano naturalmente: $f$ è **iniettiva** se $f(a_1)=f(a_2)\Rightarrow a_1=a_2$; è **suriettiva** su $B$ se per ogni $b\in B$ esiste $a\in A$ con $f(a)=b$.

### Relazioni di equivalenza

Una relazione $R$ su un insieme $A$ (cioè $R\subseteq A\times A$) è una **relazione di equivalenza** se soddisfa tre proprietà:

- **riflessiva**: $a\,R\,a$ per ogni $a$;
- **simmetrica**: $a\,R\,b\Rightarrow b\,R\,a$;
- **transitiva**: $a\,R\,b\wedge b\,R\,c\Rightarrow a\,R\,c$.

L'esempio più familiare è l'**uguaglianza** stessa, ma non è l'unico: "avere lo stesso resto nella divisione per $n$" (congruenza modulo $n$) è una relazione di equivalenza su $\mathbb Z$; "avere la stessa cardinalità" (che definiremo nel notebook 10.5) sarà una relazione di equivalenza sugli insiemi. Una relazione di equivalenza **partiziona** $A$ in **classi di equivalenza**: sottoinsiemi disgiunti la cui unione è tutto $A$, ciascuno formato dagli elementi "equivalenti" tra loro. Questa idea di raggruppare oggetti diversi ma "equivalenti secondo un criterio" è precisamente ciò che faremo nel notebook 10.5, dove diremo che due insiemi hanno "la stessa cardinalità" se esiste una biiezione tra loro — una relazione di equivalenza sugli insiemi.

### Relazioni d'ordine

Una relazione $R$ su $A$ è una **relazione d'ordine (parziale)** se è riflessiva, **antisimmetrica** ($a\,R\,b\wedge b\,R\,a\Rightarrow a=b$) e transitiva. L'esempio canonico è $\le$ su $\mathbb R$, già usato assiomaticamente nel notebook 5.1. Se inoltre per ogni $a,b\in A$ vale $a\,R\,b$ oppure $b\,R\,a$ (due elementi qualsiasi sono sempre confrontabili), l'ordine si dice **totale**: è il caso di $\le$ su $\mathbb R$, ma non, per esempio, dell'inclusione $\subseteq$ tra insiemi (due insiemi generici non sono necessariamente confrontabili per inclusione). Questa nozione di ordine totale è quella su cui poggia l'assioma di completezza del notebook 5.2 e la sua versione rigorosa nel notebook 10.9: senza un ordine totale, parlare di "estremo superiore" non avrebbe senso.

````python
# Verifichiamo computazionalmente le proprietà di una relazione di equivalenza:
# la congruenza modulo 3 su {0, 1, ..., 8}

A = list(range(9))

def R(a, b):
    return (a - b) % 3 == 0

def riflessiva(A, R):
    return all(R(a, a) for a in A)

def simmetrica(A, R):
    return all((not R(a, b)) or R(b, a) for a in A for b in A)

def transitiva(A, R):
    return all((not (R(a, b) and R(b, c))) or R(a, c)
               for a in A for b in A for c in A)

print("Riflessiva?", riflessiva(A, R))
print("Simmetrica?", simmetrica(A, R))
print("Transitiva?", transitiva(A, R))

# Le classi di equivalenza (dovrebbero essere 3: resto 0, 1, 2)
classi = {}
for a in A:
    resto = a % 3
    classi.setdefault(resto, []).append(a)

print("\nClassi di equivalenza (partizione di A):", classi)
print("L'unione delle classi ricopre A?", sorted(sum(classi.values(), [])) == A)
````

## Esercizio svolto e commentato

**Problema:** dimostra che la relazione $R$ su $\mathbb R$ definita da $x\,R\,y \iff x-y\in\mathbb Z$ (cioè $x$ e $y$ differiscono per un intero) è una relazione di equivalenza.

**Ragionamento:**

**Riflessività**: per ogni $x\in\mathbb R$, $x-x=0$, e $0\in\mathbb Z$. Quindi $x\,R\,x$ per ogni $x$: riflessiva.

**Simmetria**: supponiamo $x\,R\,y$, cioè $x-y=k$ per qualche $k\in\mathbb Z$. Allora $y-x=-k$, e $-k\in\mathbb Z$ (l'insieme $\mathbb Z$ è chiuso rispetto all'opposto, notebook 0.2). Quindi $y\,R\,x$: simmetrica.

**Transitività**: supponiamo $x\,R\,y$ e $y\,R\,z$, cioè $x-y=k_1\in\mathbb Z$ e $y-z=k_2\in\mathbb Z$. Sommando le due uguaglianze: $(x-y)+(y-z)=k_1+k_2$, cioè $x-z=k_1+k_2$. Poiché $\mathbb Z$ è chiuso rispetto alla somma, $k_1+k_2\in\mathbb Z$, quindi $x\,R\,z$: transitiva.

**Conclusione:** $R$ è una relazione di equivalenza. **Cosa sono le classi di equivalenza?** Fissato $x\in\mathbb R$, la classe di $x$ è $\{y\in\mathbb R: y-x\in\mathbb Z\}=\{x+k: k\in\mathbb Z\}$ — tutti i numeri reali che differiscono da $x$ per un intero. Ogni classe ha esattamente un rappresentante nell'intervallo $[0,1)$ (la parte frazionaria di $x$): questa relazione è precisamente ciò che identifica $\mathbb R$ con l'insieme delle "fasi" di un fenomeno periodico, un'idea che ritroveremo, in forma diversa, quando parleremo di funzioni periodiche e serie di Fourier (già introdotte nel notebook 8.9).

````python
import sympy as sp

def R(x, y):
    return sp.nsimplify(x - y).is_integer

casi = [
    (sp.Rational(7, 2), sp.Rational(3, 2)),   # differiscono di 2, intero
    (sp.pi, sp.pi + 3),                        # differiscono di 3, intero
    (sp.Rational(1, 3), sp.Rational(2, 3)),   # differiscono di 1/3, non intero
]

for x, y in casi:
    print(f"x={x}, y={y}: x - y = {sp.nsimplify(x-y)}, x R y? {R(x, y)}")
````

**3.** Su tutto $\mathbb R$: $f$ non è iniettiva, perché $f(-1)=f(1)=1$ ma $-1\ne1$ (contraddice la definizione: $f(a_1)=f(a_2)$ non implica $a_1=a_2$). Iniettiva su $[0,+\infty)$.

**4.** Supponiamo $[a]\cap[b]\ne\emptyset$: esiste $c$ con $c\in[a]$ e $c\in[b]$, cioè $a\,R\,c$ e $b\,R\,c$. Per simmetria, $c\,R\,b$. Per transitività applicata a $a\,R\,c$ e $c\,R\,b$: $a\,R\,b$. Ora, se $x\in[a]$ (cioè $a\,R\,x$), per simmetria $x\,R\,a$, e per transitività con $a\,R\,b$ si ottiene $x\,R\,b$, cioè $x\in[b]$: quindi $[a]\subseteq[b]$. ## Riepilogo finale

- Una **funzione** è, rigorosamente, un sottoinsieme di $A\times B$ che soddisfa totalità e unicità — non più una "regola" informale.
- Una **relazione di equivalenza** (riflessiva, simmetrica, transitiva) partiziona un insieme in **classi di equivalenza**, che sono sempre disgiunte o coincidenti.
- Una **relazione d'ordine** (riflessiva, antisimmetrica, transitiva) formalizza il confronto tra elementi; è **totale** se ogni coppia è confrontabile, come $\le$ su $\mathbb R$.
- Questi strumenti — in particolare le classi di equivalenza — sono il fondamento su cui costruiremo, nel prossimo notebook, la nozione di "avere lo stesso numero di elementi" per insiemi infiniti.
