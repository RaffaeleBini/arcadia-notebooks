## Teoria, passo dopo passo

### La derivata come limite: la definizione formale

$$f'(x_0)=\lim_{h\to0}\frac{f(x_0+h)-f(x_0)}h \quad\stackrel{\text{def}}{\Longleftrightarrow}\quad \forall\varepsilon>0\,\exists\delta>0:\;0<|h|<\delta\Rightarrow\left|\frac{f(x_0+h)-f(x_0)}h-f'(x_0)\right|<\varepsilon$$

È semplicemente la definizione ε-δ del notebook 10.8, applicata alla funzione (variabile $h$) $g(h)=\dfrac{f(x_0+h)-f(x_0)}h$, con limite per $h\to0$.

### Derivabilità implica continuità (dimostrazione completa)

**Teorema.** Se $f$ è derivabile in $x_0$, allora $f$ è continua in $x_0$.

**Dimostrazione.** Vogliamo $f(x_0+h)\to f(x_0)$ per $h\to0$ (definizione di continuità, riscritta con $x=x_0+h$). Scriviamo, per $h\ne0$:

$$f(x_0+h)-f(x_0)=\frac{f(x_0+h)-f(x_0)}h\cdot h$$

(un'identità puramente algebrica, valida perché stiamo solo moltiplicando e dividendo per $h\ne0$). Per ipotesi, $\dfrac{f(x_0+h)-f(x_0)}h\to f'(x_0)$ per $h\to0$ (definizione di derivabilità); inoltre banalmente $h\to0$. Per la regola del prodotto dell'algebra dei limiti (notebook 10.10, applicabile ai limiti di funzione con lo stesso schema visto per le successioni): $\dfrac{f(x_0+h)-f(x_0)}h\cdot h\to f'(x_0)\cdot0=0$. Dunque $f(x_0+h)-f(x_0)\to0$, cioè $f(x_0+h)\to f(x_0)$: $f$ è continua in $x_0$. $\blacksquare$

**Perché il viceversa è falso**: lo abbiamo già visto nel notebook 10.7 con $f(x)=|x|$ in $x=0$ — continua ma non derivabile. La dimostrazione qui sopra chiarisce *perché* l'implicazione va in un solo verso: la continuità richiede solo che $f(x_0+h)-f(x_0)\to0$, mentre la derivabilità richiede il molto più forte "$f(x_0+h)-f(x_0)$ tende a $0$ **alla stessa velocità di $h$**" (il rapporto ha limite finito) — $|x|$ soddisfa la prima richiesta ma non la seconda, perché il rapporto incrementale oscilla tra $+1$ e $-1$ senza stabilizzarsi.

### Teorema di Fermat (punti critici interni)

**Lemma (Fermat).** Se $f$ è derivabile in $c$ e $c$ è un punto di massimo (o minimo) **locale interno**, allora $f'(c)=0$.

**Dimostrazione.** Sia $c$ un massimo locale: esiste un intorno $(c-\delta,c+\delta)$ dove $f(x)\le f(c)$. Per $h>0$ piccolo: $\dfrac{f(c+h)-f(c)}h\le0$ (numeratore $\le0$, denominatore $>0$), quindi il limite destro soddisfa $f'(c)\le0$ (i limiti conservano le disuguaglianze deboli, un fatto dimostrabile per assurdo con la definizione ε-δ). Per $h<0$ piccolo: $\dfrac{f(c+h)-f(c)}h\ge0$ (numeratore $\le0$, denominatore $<0$), quindi il limite sinistro soddisfa $f'(c)\ge0$. Essendo $f$ derivabile in $c$, il limite destro e sinistro coincidono con $f'(c)$: $f'(c)\le0$ e $f'(c)\ge0$ insieme danno $f'(c)=0$. $\blacksquare$

### Teorema di Rolle e teorema di Lagrange (dimostrazione completa)

**Teorema (Rolle).** Se $f$ è continua su $[a,b]$, derivabile su $(a,b)$, e $f(a)=f(b)$, allora esiste $c\in(a,b)$ con $f'(c)=0$.

**Dimostrazione.** Per il teorema di Weierstrass (notebook 10.12, applicabile perché $f$ è continua su $[a,b]$ compatto), $f$ ammette massimo $M$ e minimo $m$ su $[a,b]$. Se $M=m$, $f$ è costante, quindi $f'\equiv0$ su $(a,b)$ e qualunque $c$ funziona. Altrimenti $M>m$; poiché $f(a)=f(b)$, almeno uno tra $M$ e $m$ è raggiunto in un punto $c$ **interno** ad $(a,b)$ (non può essere raggiunto solo agli estremi, dato che $f(a)=f(b)$ e $M\ne m$ implicano che il valore diverso da $f(a)$ sia assunto altrove). In questo $c$, per il teorema di Fermat, $f'(c)=0$. $\blacksquare$

**Teorema (Lagrange).** Se $f$ è continua su $[a,b]$ e derivabile su $(a,b)$, allora esiste $c\in(a,b)$ con $f'(c)=\dfrac{f(b)-f(a)}{b-a}$.

**Dimostrazione.** Costruiamo una funzione ausiliaria che "raddrizza" la secante, riducendoci al caso di Rolle:

$$g(x)=f(x)-f(a)-\frac{f(b)-f(a)}{b-a}(x-a)$$

$g$ è continua su $[a,b]$ e derivabile su $(a,b)$ (differenza di funzioni continue/derivabili, con l'ultimo termine lineare quindi derivabile ovunque). Verifichiamo $g(a)=g(b)$: $g(a)=f(a)-f(a)-0=0$; $g(b)=f(b)-f(a)-\dfrac{f(b)-f(a)}{b-a}(b-a)=f(b)-f(a)-(f(b)-f(a))=0$. Dunque $g(a)=g(b)=0$: per il teorema di Rolle appena dimostrato, esiste $c\in(a,b)$ con $g'(c)=0$. Calcolando $g'(x)=f'(x)-\dfrac{f(b)-f(a)}{b-a}$, la condizione $g'(c)=0$ dà esattamente $f'(c)=\dfrac{f(b)-f(a)}{b-a}$. $\blacksquare$

Questa dimostrazione mostra che Lagrange **non** è un fatto indipendente da Rolle, ma una sua conseguenza diretta, ottenuta con un trucco algebrico (sottrarre la retta secante) — esattamente come suggeriva l'immagine geometrica del notebook 5.11, ora tradotta in formule.

````python
import sympy as sp

x, a, b = sp.symbols('x a b')
f = x**3 - 3*x  # esempio concreto

a_val, b_val = sp.Integer(-1), sp.Integer(2)
f_expr = f.subs(x, x)  # solo per chiarezza

# Retta secante tra (a, f(a)) e (b, f(b))
pendenza_secante = (f.subs(x, b_val) - f.subs(x, a_val)) / (b_val - a_val)
print("Pendenza della secante tra a e b:", pendenza_secante)

# Troviamo c tale che f'(c) = pendenza_secante (teorema di Lagrange)
f_prime = sp.diff(f, x)
soluzioni_c = sp.solve(sp.Eq(f_prime, pendenza_secante), x)
soluzioni_nell_intervallo = [c for c in soluzioni_c if a_val < c < b_val]
print("Punti c che soddisfano Lagrange (f'(c) = pendenza secante):", soluzioni_c)
print("Di questi, quelli effettivamente in (a,b):", soluzioni_nell_intervallo)
````

## Esercizio svolto e commentato

**Problema:** usa il teorema di Lagrange per dimostrare che se $f'(x)=0$ per ogni $x$ in un intervallo $I$, allora $f$ è **costante** su $I$ (un fatto usato implicitamente ogni volta che si dice "due primitive differiscono per una costante", notebook 5.16).

**Ragionamento:**

**Passo 1 — fissiamo due punti arbitrari.** Siano $x_1<x_2$ due punti qualsiasi di $I$. Vogliamo dimostrare $f(x_1)=f(x_2)$.

**Passo 2 — applichiamo Lagrange sull'intervallo $[x_1,x_2]$.** $f$ è derivabile (quindi continua, per il teorema dimostrato sopra) su tutto $I\supseteq[x_1,x_2]$, quindi le ipotesi di Lagrange sono soddisfatte. Esiste $c\in(x_1,x_2)$ con:

$$f'(c)=\frac{f(x_2)-f(x_1)}{x_2-x_1}$$

**Passo 3 — usiamo l'ipotesi $f'\equiv0$.** Poiché $c\in I$ e $f'(x)=0$ per ogni $x\in I$, in particolare $f'(c)=0$. Sostituendo:

$$0=\frac{f(x_2)-f(x_1)}{x_2-x_1}$$

**Passo 4 — risolviamo.** Poiché $x_2-x_1\ne0$, l'unico modo perché la frazione sia $0$ è che il numeratore sia $0$: $f(x_2)-f(x_1)=0$, cioè $f(x_1)=f(x_2)$.

**Conclusione:** per ogni $x_1,x_2\in I$, $f(x_1)=f(x_2)$: $f$ è costante su $I$. **Perché serve che $I$ sia un intervallo** (e non, per esempio, un'unione di due intervalli separati): la dimostrazione applica Lagrange sul segmento $[x_1,x_2]$, che deve restare **interamente dentro** $I$ perché $f$ sia derivabile su tutto quel segmento — su un dominio con "buchi", due punti in componenti diverse potrebbero avere derivata sempre nulla senza che la funzione sia globalmente costante (basta pensare a $f(x)=1$ su $(0,1)$ e $f(x)=2$ su $(2,3)$: $f'\equiv0$ ovunque sia definita, ma $f$ non è costante sull'unione).

````python
import sympy as sp

x = sp.symbols('x')

# Verifica: una funzione con derivata identicamente nulla su un intervallo e' costante
f = sp.Function('f')
c1 = sp.symbols('c1')
f_costante = c1  # se f' = 0, f deve essere una costante

derivata = sp.diff(f_costante, x)
print("Derivata di una funzione costante c1:", derivata, "(sempre 0, come richiesto)")

# Controesempio sulla necessita' dell'intervallo: dominio con un "buco"
def f_a_pezzi(x_val):
    if x_val < 1.5:
        return 1
    else:
        return 2

print("\nControesempio (dominio NON un intervallo):")
for x_val in [0.5, 1.0, 2.0, 2.5]:
    print(f"f({x_val}) = {f_a_pezzi(x_val)}  (derivata nulla localmente, ma f non è costante globalmente)")
````

## Riepilogo finale

- Con la stessa identità algebrica $f(x_0+h)-f(x_0)=\dfrac{f(x_0+h)-f(x_0)}h\cdot h$, si dimostra rigorosamente che **derivabile $\Rightarrow$ continua**.
- Il **teorema di Fermat** (derivata nulla nei punti critici interni) segue confrontando i limiti destro e sinistro del rapporto incrementale in un massimo/minimo locale.
- Il **teorema di Rolle** segue da Fermat applicato al massimo o minimo garantito da Weierstrass; il **teorema di Lagrange** segue da Rolle applicato a una funzione ausiliaria che sottrae la retta secante.
- La conseguenza più usata di Lagrange: **derivata nulla ovunque su un intervallo $\Rightarrow$ funzione costante**, il fondamento di ogni discorso su primitive e studio di funzione.
