# 10.9 — Completezza (rigorosa)

**Prerequisiti:** 5.2 Completezza dei reali, 10.5 Cardinalità, 10.6 Dimostrazioni dirette, per contrapposizione, per assurdo (approfondimento).


## Perché ci interessa

Nel notebook 5.2 abbiamo detto, in modo intuitivo, che $\mathbb R$ "non ha buchi" mentre $\mathbb Q$ sì — l'esempio classico era che $\sqrt2$ non ha posto in $\mathbb Q$ (dimostrato rigorosamente nel notebook 10.6) ma esiste in $\mathbb R$. Questa proprietà, chiamata **completezza**, è l'unica cosa che distingue algebricamente $\mathbb R$ da $\mathbb Q$ (entrambi sono campi ordinati, nel linguaggio del notebook 0.5 e 10.3) — ed è il fondamento su cui poggia **ogni** teorema di esistenza dell'analisi: l'esistenza del massimo di una funzione continua (notebook 10.12), la convergenza di una successione di Cauchy (notebook 10.10), l'esistenza dell'integrale di Riemann (notebook 10.15). Questo notebook la formalizza con precisione assoluta, come **assioma**, e ne deriva le prime conseguenze rigorose.

## Teoria, passo dopo passo

### Maggioranti, estremo superiore: definizioni precise

Sia $A\subseteq\mathbb R$ non vuoto. Un numero $M\in\mathbb R$ è un **maggiorante** di $A$ se $a\le M$ per ogni $a\in A$. Se $A$ ha almeno un maggiorante, si dice **limitato superiormente**. L'**estremo superiore** di $A$, indicato $\sup A$, è il **più piccolo** dei maggioranti — rigorosamente: $s=\sup A$ se e solo se

1. $s$ è un maggiorante di $A$ ($a\le s$ per ogni $a\in A$);
2. per ogni $\varepsilon>0$, esiste $a\in A$ con $a>s-\varepsilon$ (nessun numero più piccolo di $s$ è un maggiorante).

La condizione 2 è la formalizzazione precisa di "il **più piccolo**" maggiorante: se un numero $s-\varepsilon$ più piccolo di $s$ fosse anch'esso maggiorante, per definizione **tutti** gli elementi di $A$ dovrebbero essere $\le s-\varepsilon$ — ma la condizione 2 esibisce un elemento che lo supera, contraddicendo questo. Analogamente si definiscono minorante ed **estremo inferiore** $\inf A$ (il più grande dei minoranti).

### L'assioma di completezza

**Assioma (completezza di $\mathbb R$).** Ogni sottoinsieme $A\subseteq\mathbb R$ non vuoto e limitato superiormente ammette estremo superiore $\sup A\in\mathbb R$.

Questo è l'assioma che $\mathbb Q$ **non** soddisfa: l'insieme $A=\{q\in\mathbb Q:q^2<2\}$ è limitato superiormente (per esempio da $2$) ma non ha estremo superiore in $\mathbb Q$ — il candidato naturale sarebbe $\sqrt2$, che non è razionale (notebook 10.6). $\mathbb R$ è, per assioma, costruito in modo da colmare esattamente questo tipo di lacuna: ogni insieme limitato "punta verso" un numero reale specifico, anche quando quel numero non è raggiungibile con operazioni algebriche elementari sui razionali.

### Conseguenza 1: la proprietà archimedea

**Teorema.** Per ogni $x\in\mathbb R$, esiste $n\in\mathbb N$ con $n>x$ ($\mathbb N$ non è limitato superiormente in $\mathbb R$).

**Dimostrazione (per assurdo).** Supponiamo che $\mathbb N$ sia limitato superiormente. Per l'assioma di completezza, esiste $s=\sup\mathbb N\in\mathbb R$. Per la condizione 2 della definizione di estremo superiore, con $\varepsilon=1$, esiste $n\in\mathbb N$ con $n>s-1$, cioè $n+1>s$. Ma $n+1\in\mathbb N$ (chiusura di $\mathbb N$ rispetto al successore, notebook 0.1, già usata nel notebook 10.6), quindi $n+1$ è un elemento di $\mathbb N$ che **supera** $s=\sup\mathbb N$ — contraddicendo che $s$ fosse un maggiorante. Contraddizione: $\mathbb N$ non è limitato superiormente. $\blacksquare$

**Perché questo teorema, apparentemente ovvio, richiede l'assioma di completezza:** in un ipotetico campo ordinato "esotico" con elementi infinitamente grandi (più grandi di ogni numero naturale), la proprietà archimedea fallirebbe pur mantenendo tutte le regole algebriche dell'ordine. È **specificamente** la completezza a escludere questa possibilità in $\mathbb R$.

### Conseguenza 2: densità di $\mathbb Q$ in $\mathbb R$

**Teorema.** Per ogni $a,b\in\mathbb R$ con $a<b$, esiste $q\in\mathbb Q$ con $a<q<b$.

**Dimostrazione.** Per la proprietà archimedea, esiste $n\in\mathbb N$ con $n>\dfrac1{b-a}$, cioè $\dfrac1n<b-a$ (moltiplicando e dividendo, notebook 5.1). Consideriamo l'insieme dei multipli interi di $\dfrac1n$ che superano $a$: per la proprietà archimedea (applicata a $na$), esiste un **primo** intero $m$ con $\dfrac mn>a$ (l'insieme $\{k\in\mathbb Z:k/n>a\}$ è non vuoto per Archimede ed è limitato inferiormente, quindi ha minimo, un fatto che discende dal buon ordinamento di $\mathbb Z$). Per minimalità di $m$, $\dfrac{m-1}n\le a$, quindi $\dfrac mn\le a+\dfrac1n<a+(b-a)=b$. Dunque $q=\dfrac mn$ soddisfa $a<q<b$. $\blacksquare$

### Conseguenza 3: il teorema degli intervalli incapsulati

**Teorema.** Sia $I_1\supseteq I_2\supseteq I_3\supseteq\cdots$ una successione di intervalli chiusi e limitati $I_n=[a_n,b_n]$, incapsulati (ciascuno contenuto nel precedente). Allora $\bigcap_{n=1}^\infty I_n\ne\emptyset$.

**Dimostrazione.** L'incapsulamento implica $a_1\le a_2\le\cdots\le b_2\le b_1$ (ogni $a_n$ è un minorante per tutti i $b_m$ successivi). L'insieme $A=\{a_n:n\in\mathbb N\}$ è limitato superiormente (per esempio da $b_1$), quindi per l'assioma di completezza esiste $s=\sup A$. Per costruzione $s\ge a_n$ per ogni $n$ (definizione di maggiorante). Inoltre, ogni $b_m$ è un maggiorante di $A$ (perché $a_n\le b_m$ per ogni $n$, come si verifica separando i casi $n\le m$ e $n>m$ usando l'incapsulamento), quindi $s\le b_m$ per ogni $m$ (perché $s$ è il **più piccolo** dei maggioranti). Dunque $a_n\le s\le b_n$ per ogni $n$: $s\in I_n$ per ogni $n$, cioè $s\in\bigcap_nI_n$. $\blacksquare$

Questo teorema, apparentemente tecnico, sarà l'ingrediente chiave della dimostrazione del teorema di Bolzano-Weierstrass nel notebook 10.10, tramite il metodo di bisezione.

````python
import sympy as sp

# Verifica numerica dell'estremo superiore di A = {q in Q : q^2 < 2}: dovrebbe approssimare sqrt(2)
def sup_approssimato(precisione_decimali):
    x = sp.Rational(1)
    passo = sp.Rational(1)
    for _ in range(precisione_decimali * 4):  # raffiniamo iterativamente
        while (x + passo)**2 < 2:
            x += passo
        passo /= 10
    return x

approssimazione = sup_approssimato(6)
print("Estremo superiore approssimato di {q in Q : q^2 < 2}:", float(approssimazione))
print("sqrt(2) =", float(sp.sqrt(2)))
print("Differenza:", float(sp.sqrt(2)) - float(approssimazione))

# Verifica della proprietà archimedea: per ogni x, esiste n > x
import math
for x_val in [10, 1000, 1e9]:
    n = math.floor(x_val) + 1
    print(f"x={x_val}: n={n} soddisfa n > x? {n > x_val}")
````

## Esercizio svolto e commentato

**Problema:** dimostra che $\sup A=1$ per $A=\left\{1-\dfrac1n:n\in\mathbb N_{>0}\right\}=\left\{0,\dfrac12,\dfrac23,\dfrac34,\ldots\right\}$.

**Ragionamento:**

Dobbiamo verificare le due condizioni della definizione di estremo superiore.

**Condizione 1 — $1$ è un maggiorante.** Per ogni $n\in\mathbb N_{>0}$, $\dfrac1n>0$, quindi $1-\dfrac1n<1$. Dunque ogni elemento di $A$ è $<1\le1$: $1$ è un maggiorante.

**Condizione 2 — nessun numero più piccolo di $1$ è un maggiorante.** Fissiamo $\varepsilon>0$ arbitrario. Dobbiamo trovare un elemento di $A$ maggiore di $1-\varepsilon$, cioè un $n$ con $1-\dfrac1n>1-\varepsilon$, cioè $\dfrac1n<\varepsilon$, cioè $n>\dfrac1\varepsilon$. Per la proprietà archimedea (dimostrata nella teoria), un tale $n\in\mathbb N$ **esiste sempre**, qualunque sia $\varepsilon>0$.

**Conclusione:** entrambe le condizioni sono soddisfatte: $\sup A=1$. **Osservazione importante:** $1\notin A$ (nessun termine della successione $1-1/n$ raggiunge esattamente $1$): l'estremo superiore, a differenza del massimo, **non deve appartenere** all'insieme. Questa distinzione tra $\sup$ (che esiste sempre, per la completezza, quando l'insieme è limitato) e massimo (che può non esistere, come qui) è cruciale, e riemergerà nel notebook 10.12 quando parleremo di funzioni continue che raggiungono effettivamente il proprio massimo — un fatto **non banale**, che richiede ipotesi aggiuntive (la compattezza del dominio, notebook 10.11).

````python
from fractions import Fraction

A = [1 - Fraction(1, n) for n in range(1, 30)]

print("Primi termini di A = {1 - 1/n}:", [float(a) for a in A[:8]])
print("Massimo osservato nei primi 30 termini:", float(max(A)))
print("1 appartiene ad A per qualche n testato?", any(a == 1 for a in A))

# Verifica della condizione 2: per ogni epsilon, esiste un elemento di A oltre 1-epsilon
def testimone_esiste(epsilon, n_max=100000):
    for n in range(1, n_max):
        if 1 - 1.0/n > 1 - epsilon:
            return n
    return None

for eps in [0.1, 0.001, 0.0001]:
    n_trovato = testimone_esiste(eps)
    print(f"epsilon={eps}: primo n che soddisfa la condizione 2 -> n={n_trovato}")
````

## Esercizi da risolvere

1. Dimostra che $\inf\left\{\dfrac1n:n\in\mathbb N_{>0}\right\}=0$ (usando la proprietà archimedea, in modo analogo all'esercizio svolto).
2. Dimostra che se $A\subseteq B$, entrambi non vuoti e limitati superiormente, allora $\sup A\le\sup B$.
3. Dimostra che $\sup(A\cup B)=\max(\sup A,\sup B)$ per $A,B$ non vuoti e limitati superiormente.
4. *(più impegnativo)* Dimostra che $\sqrt2=\sup\{q\in\mathbb Q:q^2<2\}$ esiste come numero reale (usa l'assioma di completezza per garantire l'esistenza di $s=\sup A$, poi dimostra per assurdo che né $s^2<2$ né $s^2>2$ sono possibili, concludendo $s^2=2$).

**1.** $0$ è un minorante ovvio ($1/n>0$ per ogni $n$). Per la condizione 2 (analoga a quella del $\sup$, con i ruoli scambiati): fissato $\varepsilon>0$, per Archimede esiste $n$ con $n>1/\varepsilon$, cioè $1/n<\varepsilon=0+\varepsilon$: esiste un elemento dell'insieme minore di $0+\varepsilon$, quindi nessun numero $>0$ è un minorante. Dunque $\inf=0$.

**2.** Sia $s=\sup B$. Per ogni $a\in A$, poiché $A\subseteq B$, si ha $a\in B$, quindi $a\le s$ (perché $s$ è un maggiorante di $B$). Dunque $s$ è anche un maggiorante di $A$. Ma $\sup A$ è il **più piccolo** dei maggioranti di $A$, quindi $\sup A\le s=\sup B$.

**3.** Sia $M=\max(\sup A,\sup B)$. $M$ è un maggiorante sia di $A$ (perché $M\ge\sup A$) sia di $B$ (perché $M\ge\sup B$), quindi di $A\cup B$. Per la minimalità, dobbiamo verificare che nessun numero minore di $M$ sia un maggiorante di $A\cup B$: supponiamo, senza perdita di generalità, $M=\sup A$. Fissato $\varepsilon>0$, per la condizione 2 applicata ad $A$, esiste $a\in A\subseteq A\cup B$ con $a>M-\varepsilon$. Quindi $M$ soddisfa entrambe le condizioni: $\sup(A\cup B)=M=\max(\sup A,\sup B)$.

**4.** Per l'assioma di completezza, $s=\sup A$ esiste, con $A=\{q\in\mathbb Q:q^2<2\}$ ($A$ è non vuoto, es. $1\in A$, ed è limitato superiormente da $2$). Supponiamo per assurdo $s^2<2$: si può mostrare (con un calcolo algebrico) che per $h>0$ abbastanza piccolo, $(s+h)^2<2$ ancora, quindi $s+h\in A$ con $s+h>s$, contraddicendo che $s$ fosse un maggiorante. Supponiamo per assurdo $s^2>2$: analogamente si trova $h>0$ tale che $(s-h)^2>2$, quindi $s-h$ è un maggiorante di $A$ più piccolo di $s$ (perché ogni $q\in A$ ha $q^2<2<(s-h)^2$, quindi $q<s-h$ dato che entrambi positivi), contraddicendo la minimalità di $s$. Restando esclusi $s^2<2$ e $s^2>2$, per tricotomia dell'ordine su $\mathbb R$ deve valere $s^2=2$: $s=\sqrt2$ esiste come numero reale, definito esattamente come estremo superiore di un insieme di razionali.

Verifica computazionale (esercizio 4, approssimazione numerica di s = sup A):
```python
def approssima_sup(iterazioni=50):
    basso, alto = 1.0, 2.0
    for _ in range(iterazioni):
        medio = (basso + alto) / 2
        if medio**2 < 2:
            basso = medio
        else:
            alto = medio
    return basso

s = approssima_sup()
print("Approssimazione di sup{q in Q : q^2 < 2}:", s)
print("s^2 =", s**2, " (atteso vicino a 2)")
```

## Riepilogo finale

- L'**estremo superiore** $\sup A$ è il più piccolo maggiorante: formalmente, un maggiorante tale che nessun numero più piccolo lo sia (condizione ε).
- L'**assioma di completezza** garantisce che ogni insieme non vuoto e limitato superiormente ha estremo superiore in $\mathbb R$ — la proprietà che $\mathbb Q$ non ha.
- Da questo assioma discendono, con dimostrazioni rigorose, la **proprietà archimedea** ($\mathbb N$ non limitato in $\mathbb R$), la **densità di $\mathbb Q$** in $\mathbb R$, e il **teorema degli intervalli incapsulati**.
- $\sup A$ non deve appartenere ad $A$ (a differenza del massimo): una distinzione che tornerà cruciale nel teorema di Weierstrass (notebook 10.12).
