## Teoria, passo dopo passo

### Dimostrazione diretta

Per dimostrare $p\Rightarrow q$ **direttamente**, si assume $p$ vera e si costruisce una catena di implicazioni $p\Rightarrow p_1\Rightarrow p_2\Rightarrow\cdots\Rightarrow q$, ciascuna giustificata da un fatto già noto. È la tecnica più naturale, usata nella maggior parte degli esercizi svolti finora nel corso (per esempio, l'esercizio del notebook 10.4 sulla disuguaglianza di Bernoulli).

### Dimostrazione per contrapposizione

Nel notebook 10.1 abbiamo dimostrato che $p\Rightarrow q$ è **equivalente** alla sua contronominale $\neg q\Rightarrow\neg p$. Questo significa che, per dimostrare $p\Rightarrow q$, è **logicamente lecito** dimostrare invece $\neg q\Rightarrow\neg p$: si assume che la tesi sia falsa, e si deduce che anche l'ipotesi lo è. È utile quando negare la tesi produce un'ipotesi di lavoro più maneggevole della tesi diretta (per esempio, quando la tesi è un'affermazione di non uguaglianza o di non appartenenza).

### Dimostrazione per assurdo

Per dimostrare che una proposizione $p$ è vera, si **assume $\neg p$** (che sia falsa) e si deduce, con una catena di implicazioni corrette, una **contraddizione** — una proposizione della forma $r\wedge\neg r$, che sappiamo essere sempre falsa (notebook 10.1, esercizio 2). Poiché da un'ipotesi vera non si può mai dedurre correttamente il falso, l'ipotesi $\neg p$ deve essere falsa, cioè $p$ è vera.

**Perché è logicamente valida**: se $\neg p\Rightarrow(r\wedge\neg r)$ è una catena di implicazioni corrette, allora $\neg p\Rightarrow(r\wedge\neg r)$ è vera; ma $(r\wedge\neg r)$ è sempre falsa, quindi (per la tabella di verità di $\Rightarrow$, notebook 10.1: un'implicazione vera con conseguente falso richiede l'antecedente falso) $\neg p$ deve essere falsa, cioè $p$ è vera. La dimostrazione per assurdo è, in un certo senso, un caso particolare della contrapposizione: dimostrare $p$ per assurdo equivale a dimostrare "$\text{Vero}\Rightarrow p$" per contrapposizione, cioè $\neg p\Rightarrow\text{Falso}$.

### Un classico: $\sqrt2$ è irrazionale (per assurdo)

**Tesi:** $\sqrt2\notin\mathbb Q$.

**Dimostrazione.** Supponiamo per assurdo $\sqrt2\in\mathbb Q$: allora $\sqrt2=p/q$ con $p,q\in\mathbb N$, $q\ne0$, e **ridotta ai minimi termini** (cioè $\gcd(p,q)=1$, sempre possibile per una frazione, notebook 0.3). Elevando al quadrato: $2=p^2/q^2$, cioè $p^2=2q^2$. Quindi $p^2$ è pari, e questo implica che $p$ stesso è pari (se $p$ fosse dispari, $p^2$ sarebbe dispari — un fatto verificabile per casi, o per contrapposizione). Scriviamo $p=2k$: sostituendo, $4k^2=2q^2$, cioè $q^2=2k^2$, quindi anche $q^2$ è pari, e per lo stesso argomento $q$ è pari. Ma allora **sia $p$ sia $q$ sono pari**, contraddicendo il fatto che $p/q$ era ridotta ai minimi termini ($\gcd(p,q)=1$, mentre 2 divide entrambi). Contraddizione: l'ipotesi $\sqrt2\in\mathbb Q$ è falsa, quindi $\sqrt2$ è irrazionale.

### Un classico: i numeri primi sono infiniti (per assurdo, Euclide)

**Tesi:** l'insieme dei numeri primi è infinito.

**Dimostrazione.** Supponiamo per assurdo che i numeri primi siano **finiti**: $p_1,p_2,\ldots,p_n$, un elenco completo. Costruiamo $N=p_1p_2\cdots p_n+1$. Per ogni $p_i$ nell'elenco, dividendo $N$ per $p_i$ si ottiene resto $1$ (perché $p_1\cdots p_n$ è divisibile per $p_i$, ma $+1$ no): quindi **nessun** $p_i$ divide $N$. Ma ogni numero naturale maggiore di 1 ha almeno un divisore primo (fatto dimostrabile per induzione forte, notebook 10.4: se $N$ è primo, divide se stesso; se non lo è, si scompone in fattori più piccoli, ciascuno con un divisore primo per ipotesi induttiva). Questo divisore primo di $N$ non può essere nessuno dei $p_i$ (appena dimostrato), quindi è un primo **non presente** nell'elenco — contraddicendo che l'elenco fosse completo. Contraddizione: i numeri primi sono infiniti.

````python
from sympy import isprime, factorint

def verifica_euclide(elenco_primi):
    N = 1
    for p in elenco_primi:
        N *= p
    N += 1
    fattori = factorint(N)
    print(f"N = {'*'.join(map(str, elenco_primi))} + 1 = {N}")
    print("Fattorizzazione di N:", fattori)
    nuovi_primi = [p for p in fattori if p not in elenco_primi]
    print("Fattori primi di N NON presenti nell'elenco originale:", nuovi_primi)
    return nuovi_primi

print("Test con un piccolo elenco 'finito' di primi:")
verifica_euclide([2, 3, 5])
print()
verifica_euclide([2, 3, 5, 7, 11, 13])
````

## Esercizio svolto e commentato

**Problema:** dimostra per contrapposizione: "se $n^2$ è pari, allora $n$ è pari" (per $n\in\mathbb N$) — il fatto usato senza dimostrazione nella prova dell'irrazionalità di $\sqrt2$.

**Ragionamento:**

**Passo 1 — identifichiamo $p$ e $q$.** $p$="$n^2$ è pari", $q$="$n$ è pari". La contronominale è $\neg q\Rightarrow\neg p$, cioè "se $n$ è **dispari**, allora $n^2$ è **dispari**".

**Passo 2 — perché scegliere la contrapposizione qui.** L'ipotesi diretta ("$n^2$ pari") è scomoda da manipolare algebricamente: non sappiamo a priori come si scompone $n$. L'ipotesi contronominale ("$n$ dispari") è invece immediatamente traducibile in una formula esplicita: $n=2k+1$ per qualche $k\in\mathbb N$ (la definizione stessa di dispari, notebook 0.2).

**Passo 3 — dimostriamo $\neg q\Rightarrow\neg p$ direttamente.** Sia $n=2k+1$. Allora $n^2=(2k+1)^2=4k^2+4k+1=2(2k^2+2k)+1$ (sviluppo del quadrato di binomio, notebook 1.2, seguito da raccoglimento a fattor comune, notebook 1.3). Poiché $2k^2+2k\in\mathbb N$, questo mostra $n^2=2m+1$ con $m=2k^2+2k$: $n^2$ è della forma "dispari", cioè $n^2$ è dispari.

**Conclusione:** abbiamo dimostrato $\neg q\Rightarrow\neg p$, che per l'equivalenza vista nel notebook 10.1 dimostra $p\Rightarrow q$: "$n^2$ pari $\Rightarrow$ $n$ pari". Questo è l'esempio scolastico più chiaro del perché la contrapposizione è preferibile alla dimostrazione diretta quando l'ipotesi originale non offre un punto di partenza algebrico immediato, mentre la sua negazione sì.

````python
def verifica_contronominale(n_max=20):
    for n in range(1, n_max + 1):
        n_pari = (n % 2 == 0)
        n_quadro_pari = (n**2 % 2 == 0)
        # Verifica che "n^2 pari" e "n pari" abbiano sempre lo stesso valore di verità
        equivalenti = (n_pari == n_quadro_pari)
        if not equivalenti:
            print(f"CONTROESEMPIO trovato per n={n}!")
            return False
    print(f"Per n=1..{n_max}: '(n pari) equivale a (n^2 pari)' verificato in ogni caso.")
    return True

verifica_contronominale()
````

Contraddizione: $\sqrt3$ è irrazionale. (La dimostrazione usa in modo essenziale che 3 è primo: per un quadrato perfetto come 4, l'argomento fallirebbe, coerentemente col fatto che $\sqrt4=2$ è razionale.)

**4.** **Passo preliminare** (per assurdo): se $q\in\mathbb Q$ e $x\notin\mathbb Q$, allora $q+x\notin\mathbb Q$. Supponiamo per assurdo $q+x\in\mathbb Q$: allora $x=(q+x)-q$ sarebbe differenza di due razionali, quindi razionale (chiusura di $\mathbb Q$ rispetto alla sottrazione, notebook 0.3) — contraddicendo $x\notin\mathbb Q$. **Passo principale**: dati $a<b$ razionali, scegliamo $n\in\mathbb N$ abbastanza grande che $\frac{\sqrt2}n<b-a$ (possibile per la proprietà archimedea di $\mathbb R$, che rivedremo rigorosamente nel notebook 10.9). Il numero $x=a+\frac{\sqrt2}n$ soddisfa $a<x<b$ (perché $0<\frac{\sqrt2}n<b-a$) ed è irrazionale per il passo preliminare (somma del razionale $a$ e dell'irrazionale $\sqrt2/n$, quest'ultimo irrazionale perché $\sqrt2$ lo è e dividerlo per un razionale non nullo non può renderlo razionale, per un argomento analogo al preliminare).

## Riepilogo finale

- La **dimostrazione diretta** costruisce una catena $p\Rightarrow p_1\Rightarrow\cdots\Rightarrow q$.
- La **dimostrazione per contrapposizione** sfrutta l'equivalenza $(p\Rightarrow q)\Leftrightarrow(\neg q\Rightarrow\neg p)$: utile quando negare la tesi dà un punto di partenza più maneggevole.
- La **dimostrazione per assurdo** nega la tesi e deduce una contraddizione $r\wedge\neg r$: se la deduzione è corretta, la tesi negata deve essere falsa.
- Due dimostrazioni classiche — irrazionalità di $\sqrt2$ e infinità dei primi, entrambe per assurdo — mostrano la potenza di questa tecnica quando un approccio diretto non è disponibile.
