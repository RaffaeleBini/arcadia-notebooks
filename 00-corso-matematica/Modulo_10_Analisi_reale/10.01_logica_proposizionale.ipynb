## Teoria, passo dopo passo

### Proposizioni e connettivi: un ripasso operativo

Una **proposizione** è un'affermazione a cui si può assegnare senza ambiguità un valore di verità, vero (V) o falso (F) (notebook 0.15). I connettivi logici fondamentali — negazione $\neg$, congiunzione $\wedge$ ("e"), disgiunzione $\vee$ ("o"), implicazione $\Rightarrow$, doppia implicazione $\Leftrightarrow$ — combinano proposizioni in nuove proposizioni, il cui valore di verità dipende **solo** dai valori di verità delle proposizioni componenti (notebook 0.16). Questa proprietà si chiama **verofunzionalità**, e permette di calcolare il valore di verità di ogni formula composta con una **tabella di verità**.

### Tautologie e contraddizioni

Una formula che è **sempre vera**, indipendentemente dal valore di verità delle proposizioni che la compongono, si chiama **tautologia**. Una formula sempre falsa si chiama **contraddizione**. Le tautologie sono importanti perché rappresentano schemi di ragionamento validi in assoluto: se una formula è una tautologia, possiamo usarla come regola di inferenza in qualunque dimostrazione, qualunque siano i contenuti delle proposizioni coinvolte.

Verifichiamo con la tabella di verità che $p\Rightarrow q$ è **equivalente** a $\neg p\vee q$ (un fatto già usato implicitamente nel notebook 0.16, ora lo dimostriamo):

| $p$ | $q$ | $p\Rightarrow q$ | $\neg p\vee q$ |
|---|---|---|---|
| V | V | V | V |
| V | F | F | F |
| F | V | V | V |
| F | F | V | V |

Le due colonne coincidono riga per riga: le formule sono **equivalenti**, cioè $(p\Rightarrow q)\Leftrightarrow(\neg p\vee q)$ è una tautologia. Questa equivalenza è la chiave per negare correttamente un'implicazione, come vedremo tra poco.

### Le leggi di De Morgan

Due tautologie fondamentali, che generalizzano quanto visto per gli insiemi nel notebook 0.13 (dove $\overline{A\cup B}=\overline A\cap\overline B$ e $\overline{A\cap B}=\overline A\cup\overline B$):

$$\neg(p\wedge q)\Leftrightarrow(\neg p\vee\neg q) \qquad\qquad \neg(p\vee q)\Leftrightarrow(\neg p\wedge\neg q)$$

Non è un caso che la struttura sia identica a quella delle leggi di De Morgan sugli insiemi: l'unione degli insiemi corrisponde logicamente a "o", l'intersezione a "e", e l'appartenenza al complementare a "non" — insiemi e logica condividono la stessa algebra di fondo.

### Negare un'implicazione

Dato che $p\Rightarrow q$ equivale a $\neg p\vee q$, la sua negazione, per De Morgan, è:

$$\neg(p\Rightarrow q)\Leftrightarrow\neg(\neg p\vee q)\Leftrightarrow(p\wedge\neg q)$$

**Cioè: l'implicazione "se $p$ allora $q$" è falsa esattamente quando $p$ è vera e $q$ è falsa.** Questo è coerente con la tabella di verità sopra (l'unica riga con $p\Rightarrow q$ falso è $p$=V, $q$=F) ma ora lo abbiamo *derivato* algebricamente, non solo *osservato* nella tabella. Useremo questo schema costantemente: per dimostrare che un'implicazione **non** vale in generale, basta trovare un caso in cui l'ipotesi è vera e la tesi è falsa — un **controesempio** (che approfondiremo nel notebook 10.7).

### Implicazione, contronominale, converso

Data $p\Rightarrow q$, si chiama **contronominale** (o contrapposta) $\neg q\Rightarrow\neg p$, e **converso** $q\Rightarrow p$. Verifichiamo con le tabelle di verità che $p\Rightarrow q$ è equivalente alla sua contronominale, ma **non** al suo converso in generale — un fatto già usato nel notebook 0.20 per introdurre la dimostrazione per contrapposizione, che ora fondiamo su basi rigorose.

````python
from itertools import product

def tavola(formula, nomi):
    """Stampa la tabella di verità di `formula` (funzione di bool -> bool) sulle variabili in `nomi`."""
    for valori in product([True, False], repeat=len(nomi)):
        ambiente = dict(zip(nomi, valori))
        riga = " ".join(f"{n}={ambiente[n]!s:5}" for n in nomi)
        print(riga, "->", formula(*valori))

def implica(p, q):
    return (not p) or q

print("Tabella di p => q:")
tavola(implica, ["p", "q"])

# Verifica che (p => q) equivale a (not p) or q per ogni combinazione
def equivalenti(f, g, n):
    return all(f(*v) == g(*v) for v in product([True, False], repeat=n))

print("\n(p=>q) equivale a (not p or q)?", equivalenti(implica, lambda p, q: (not p) or q, 2))

# Contronominale: (not q) => (not p)
def contronominale(p, q):
    return implica(not q, not p)

print("p=>q equivale alla contronominale not q => not p?", equivalenti(implica, contronominale, 2))

# Converso: q => p -- NON equivalente in generale
def converso(p, q):
    return implica(q, p)

print("p=>q equivale al converso q=>p?", equivalenti(implica, converso, 2))
````

## Esercizio svolto e commentato

**Problema:** dimostra, usando le leggi di De Morgan e l'equivalenza $p\Rightarrow q\Leftrightarrow\neg p\vee q$, che $\neg(p\Rightarrow(q\wedge r))$ equivale a $p\wedge(\neg q\vee\neg r)$.

**Ragionamento:**

**Passo 1 — riscriviamo l'implicazione interna** usando l'equivalenza vista sopra, con $q\wedge r$ al posto di $q$: $p\Rightarrow(q\wedge r)\Leftrightarrow\neg p\vee(q\wedge r)$.

**Passo 2 — neghiamo entrambi i membri**: $\neg(p\Rightarrow(q\wedge r))\Leftrightarrow\neg(\neg p\vee(q\wedge r))$.

**Passo 3 — applichiamo De Morgan** alla disgiunzione: $\neg(\neg p\vee(q\wedge r))\Leftrightarrow\neg\neg p\wedge\neg(q\wedge r)\Leftrightarrow p\wedge\neg(q\wedge r)$ (la doppia negazione $\neg\neg p$ si semplifica in $p$, una tautologia elementare che si verifica subito con la tabella di verità).

**Passo 4 — applichiamo di nuovo De Morgan**, questa volta alla congiunzione interna: $\neg(q\wedge r)\Leftrightarrow\neg q\vee\neg r$.

**Conclusione:** $\neg(p\Rightarrow(q\wedge r))\Leftrightarrow p\wedge(\neg q\vee\neg r)$.

**Perché ha senso il risultato:** "$p$ implica $q$ e $r$" è falsa quando $p$ è vera ma **almeno una** delle due conclusioni $q,r$ è falsa — esattamente ciò che dice $p\wedge(\neg q\vee\neg r)$. Questo schema — negare un'implicazione la cui tesi è una congiunzione di più condizioni — ricorrerà spesso quando negheremo definizioni con più clausole, come la continuità uniforme nel notebook 10.13.

````python
from itertools import product

def formula_originale(p, q, r):
    return not (implica(p, q and r))

def implica(p, q):
    return (not p) or q

def formula_derivata(p, q, r):
    return p and ((not q) or (not r))

uguali = all(formula_originale(*v) == formula_derivata(*v) for v in product([True, False], repeat=3))
print("Le due formule sono equivalenti per ogni combinazione di verità?", uguali)
````

## Esercizi da risolvere

1. Costruisci la tabella di verità di $p\vee(\neg p)$ e verifica che è una tautologia (il **principio del terzo escluso**).
2. Costruisci la tabella di verità di $p\wedge(\neg p)$ e verifica che è una contraddizione (il **principio di non contraddizione**).
3. Dimostra, con le leggi di De Morgan, che $\neg(p\Leftrightarrow q)$ equivale a $(p\wedge\neg q)\vee(\neg p\wedge q)$ (suggerimento: parti da $p\Leftrightarrow q\Leftrightarrow(p\Rightarrow q)\wedge(q\Rightarrow p)$).
4. *(più impegnativo)* Dimostra che l'implicazione $p\Rightarrow q$ **non** è in generale equivalente al suo converso $q\Rightarrow p$, esibendo una singola riga della tabella di verità in cui i due valori differiscono.

def implica(p, q):

def doppia_impl(p, q):

def neg_doppia_impl_attesa(p, q):

## Riepilogo finale

- Le proposizioni composte hanno un valore di verità calcolabile con **tabelle di verità**; una **tautologia** è sempre vera, una **contraddizione** sempre falsa.
- $p\Rightarrow q$ equivale a $\neg p\vee q$: da qui segue che negare un'implicazione dà $p\wedge\neg q$ — antecedente vero, conseguente falso. È lo schema di ogni controesempio.
- Le **leggi di De Morgan** ($\neg(p\wedge q)\Leftrightarrow\neg p\vee\neg q$ e viceversa) permettono di negare sistematicamente formule composte, distribuendo la negazione e scambiando $\wedge$ con $\vee$.
- $p\Rightarrow q$ è equivalente alla sua **contronominale** $\neg q\Rightarrow\neg p$, ma **non** al suo converso $q\Rightarrow p$: distinzione alla base della dimostrazione per contrapposizione.
