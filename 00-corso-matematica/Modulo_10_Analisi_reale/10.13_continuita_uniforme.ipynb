# 10.13 — Continuità uniforme

**Prerequisiti:** 10.12 Continuità (rigorosa), 10.11 Compattezza, 10.2 Quantificatori (approfondimento).


## Perché ci interessa

Nella definizione di continuità del notebook 10.12, il $\delta$ può dipendere sia da $\varepsilon$ **sia dal punto $x_0$**: una funzione può richiedere un $\delta$ molto più piccolo vicino a certi punti che vicino ad altri. Questo avrà conseguenze concrete quando, nel notebook 10.15, dimostreremo che ogni funzione continua su $[a,b]$ è integrabile: la dimostrazione richiede di controllare l'oscillazione di $f$ su **tutti** i sottointervalli di una partizione **simultaneamente**, con un singolo $\delta$ che funzioni ovunque. Questo rafforzamento della continuità — un $\delta$ **uniforme**, indipendente dal punto — è l'oggetto di questo notebook, ed è esattamente lo strumento che manca ancora per costruire una teoria rigorosa dell'integrazione.

## Teoria, passo dopo passo

### Continuità puntuale vs. continuità uniforme: l'ordine dei quantificatori

Riscriviamo la continuità di $f$ su un intero insieme $E$ esplicitando **tutti** i quantificatori (a differenza del notebook 10.12, dove $x_0$ era fissato):

$$\text{continua su }E:\quad \forall x_0\in E\,\forall\varepsilon>0\,\exists\delta>0:\;|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$$

Qui $\delta$ può dipendere sia da $\varepsilon$ sia da $x_0$ (compare **dopo** entrambi). La **continuità uniforme** scambia l'ordine tra $x_0$ e $\delta$ (esattamente il fenomeno del notebook 10.2, dove l'ordine dei quantificatori cambiava il significato):

$$\text{uniformemente continua su }E:\quad \forall\varepsilon>0\,\exists\delta>0\,\forall x_0\in E:\;|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$$

Ora $\delta$ dipende **solo** da $\varepsilon$, e deve funzionare per **ogni** $x_0\in E$ simultaneamente — la stessa distinzione tra "il testimone può dipendere dalla variabile precedente" e "il testimone deve essere universale" incontrata nel notebook 10.2. Per come i quantificatori sono annidati (esercizio 4 del notebook 10.2: $\exists y\,\forall x\,P\Rightarrow\forall x\,\exists y\,P$), la continuità uniforme **implica sempre** la continuità puntuale; il viceversa, come vedremo, è falso in generale.

### Un controesempio: $f(x)=1/x$ su $(0,1)$ è continua ma non uniformemente continua

$f(x)=1/x$ è continua su $(0,1)$ (composizione/quoziente di funzioni continue, notebook 10.12). Dimostriamo che **non** è uniformemente continua, usando la negazione della definizione (tecnica del notebook 10.2 applicata qui): dobbiamo trovare un $\varepsilon>0$ tale che **per ogni** $\delta>0$ esistano $x,x_0$ con $|x-x_0|<\delta$ ma $|f(x)-f(x_0)|\ge\varepsilon$.

Fissiamo $\varepsilon=1$. Dato un $\delta>0$ qualsiasi (per quanto piccolo), scegliamo $x_0=\delta$ e $x=\delta/2$: allora $|x-x_0|=\delta/2<\delta$, ma $|f(x)-f(x_0)|=\left|\dfrac2\delta-\dfrac1\delta\right|=\dfrac1\delta$, che per $\delta$ piccolo **supera** $\varepsilon=1$ (basta $\delta<1$). Il punto cruciale: **lo stesso $\delta$** che funziona bene lontano da $0$ diventa inadeguato vicino a $0$, dove la funzione "esplode" sempre più ripidamente — non esiste un $\delta$ unico valido ovunque su $(0,1)$.

### Teorema di Heine-Cantor: continua su un compatto ⟹ uniformemente continua

**Teorema.** Se $f$ è continua su $K$ compatto, allora $f$ è uniformemente continua su $K$.

**Dimostrazione (per assurdo, con successioni).** Supponiamo per assurdo che $f$ **non** sia uniformemente continua: esiste $\varepsilon>0$ tale che per ogni $\delta>0$ esistono $x,y\in K$ con $|x-y|<\delta$ ma $|f(x)-f(y)|\ge\varepsilon$. Scegliendo $\delta=1/n$ per ogni $n\in\mathbb N_{>0}$, otteniamo due successioni $(x_n),(y_n)$ in $K$ con $|x_n-y_n|<1/n$ ma $|f(x_n)-f(y_n)|\ge\varepsilon$ per ogni $n$.

Per compattezza sequenziale di $K$ (notebook 10.11), $(x_n)$ ha una sottosuccessione $x_{n_k}\to p\in K$. Dimostriamo che anche $y_{n_k}\to p$: per la disuguaglianza triangolare, $|y_{n_k}-p|\le|y_{n_k}-x_{n_k}|+|x_{n_k}-p|<\dfrac1{n_k}+|x_{n_k}-p|$, e entrambi i termini tendono a $0$ per $k\to\infty$ (il primo per la proprietà archimedea, il secondo per la convergenza di $x_{n_k}$).

Per la caratterizzazione sequenziale della continuità (notebook 10.12), $f(x_{n_k})\to f(p)$ **e** $f(y_{n_k})\to f(p)$ (entrambe le sottosuccessioni convergono a $p$). Quindi $f(x_{n_k})-f(y_{n_k})\to f(p)-f(p)=0$. Ma per costruzione $|f(x_{n_k})-f(y_{n_k})|\ge\varepsilon$ per **ogni** $k$: una successione che vale sempre $\ge\varepsilon$ non può tendere a $0$ (contraddizione diretta con la definizione di limite, notebook 10.10). Contraddizione: $f$ è uniformemente continua su $K$. $\blacksquare$

**Perché la compattezza è essenziale**: nel controesempio $1/x$ su $(0,1)$, l'insieme $(0,1)$ non è compatto (non chiuso, notebook 10.11) — manca proprio il punto $0$ verso cui la successione $x_n=1/n$ (e $y_n$ analoga) "vorrebbe" convergere restando dentro l'insieme. Su un compatto, questa via di fuga non esiste: ogni successione ha un punto limite **dentro** l'insieme, ed è esattamente questo che la dimostrazione sfrutta.

````python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 1/x

fig, ax = plt.subplots(figsize=(7, 4.5))
x = np.linspace(0.02, 1, 400)
ax.plot(x, f(x), color="steelblue")
ax.set_ylim(0, 20)
ax.set_title("f(x) = 1/x su (0,1): continua ma NON uniformemente continua")
plt.tight_layout()
plt.show()

# Verifica numerica del controesempio: lo stesso delta "piccolo" produce salti grandi vicino a 0
def testa_delta(delta):
    x0 = delta
    x = delta / 2
    salto = abs(f(x) - f(x0))
    print(f"delta={delta}: x0={x0}, x={x}, |x-x0|={abs(x-x0)} < delta, "
          f"ma |f(x)-f(x0)| = {salto:.2f}")

for delta in [0.5, 0.1, 0.01, 0.001]:
    testa_delta(delta)
````

## Esercizio svolto e commentato

**Problema:** dimostra direttamente (senza usare Heine-Cantor) che $f(x)=x^2$ è uniformemente continua su $[0,10]$ — un insieme **compatto**, per cui Heine-Cantor lo garantirebbe già, ma vogliamo vedere il meccanismo diretto della stima.

**Ragionamento:**

**Passo 1 — lavoro preparatorio.** $|f(x)-f(x_0)|=|x^2-x_0^2|=|x-x_0||x+x_0|$. A differenza del notebook 10.8 (dove $x_0=3$ era fissato e limitavamo $|x+3|$ localmente), qui dobbiamo limitare $|x+x_0|$ **uniformemente su tutto $[0,10]$**: poiché $x,x_0\in[0,10]$, si ha $|x+x_0|\le20$ **sempre**, indipendentemente da dove si trovino $x$ e $x_0$ nell'intervallo.

**Passo 2 — la stima diventa globale.** $|x^2-x_0^2|=|x-x_0||x+x_0|\le20|x-x_0|$, valida per **ogni** $x,x_0\in[0,10]$ simultaneamente (non serve restringere $\delta\le1$ come nel caso puntuale, perché il fattore $20$ è già un limite valido ovunque nell'intervallo).

**Passo 3 — scegliamo $\delta$.** Dato $\varepsilon>0$, vogliamo $20|x-x_0|<\varepsilon$, cioè $|x-x_0|<\varepsilon/20$: scegliamo $\delta=\varepsilon/20$, **senza alcuna dipendenza da $x_0$**.

**Conclusione:** dato $\varepsilon>0$, con $\delta=\varepsilon/20$, per **ogni** $x_0\in[0,10]$: $|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|\le20|x-x_0|<20\delta=\varepsilon$. Lo stesso $\delta$ funziona per ogni $x_0$: $f$ è uniformemente continua su $[0,10]$. **Il contrasto con $1/x$ su $(0,1)$**: lì il fattore analogo a "$20$" ($\left|\dfrac1{xx_0}\right|$ nella stima corrispondente) **non è limitato uniformemente**, perché $x,x_0$ possono avvicinarsi arbitrariamente a $0$; su $[0,10]$, invece, non c'è nessun punto "problematico" verso cui $x+x_0$ possa crescere senza limite — questo è precisamente ciò che la compattezza garantisce in generale, secondo Heine-Cantor.

````python
import numpy as np

def f(x):
    return x**2

# Verifica: lo stesso delta = eps/20 funziona per x0 vicino a 0 e per x0 vicino a 10
eps = 0.1
delta = eps / 20

for x0 in [0.01, 5.0, 9.99]:
    x = x0 + delta * 0.5  # dentro l'intorno di raggio delta
    x = min(x, 10)  # resta in [0,10]
    scarto = abs(f(x) - f(x0))
    print(f"x0={x0}: x={x:.5f}, |x-x0|={abs(x-x0):.5f} < delta={delta}, "
          f"|f(x)-f(x0)|={scarto:.6f} < eps={eps}? {scarto < eps}")
````

**1.** $|f(x)-f(x_0)|=|3x-3x_0|=3|x-x_0|$. Dato $\varepsilon>0$, scegliamo $\delta=\varepsilon/3$ (senza restrizioni, senza dipendenza da $x_0$): $|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|=3|x-x_0|<3\delta=\varepsilon$, per ogni $x_0\in\mathbb R$.

**2.** Siano $x_n=n$, $y_n=n+1/n$: $|x_n-y_n|=1/n\to0$. Ma $|f(x_n)-f(y_n)|=|n^2-(n+1/n)^2|=|{-2-1/n^2}|=2+1/n^2\to2$, che resta $\ge\varepsilon=1$ per ogni $n$ (in particolare, per $n$ grande è vicino a 2, sempre $\ge1$). Trovate due successioni con $|x_n-y_n|\to0$ ma $|f(x_n)-f(y_n)|$ che non tende a $0$ (anzi resta $\ge1$): per lo stesso schema logico usato nel controesempio della teoria, $f$ non è uniformemente continua su $\mathbb R$.

**3.** $\mathbb R$ non è **limitato** (Heine-Borel richiede chiuso **e** limitato, notebook 10.11), quindi non è compatto: il teorema di Heine-Cantor non si applica. Su $[0,10]$ (compatto) il teorema garantisce la continuità uniforme; su $\mathbb R$ (non compatto) nulla la garantisce, e infatti l'esercizio 2 mostra che effettivamente fallisce, "scappando verso l'infinito" invece che verso un punto specifico come accadeva per $1/x$ vicino a $0$.

**4.** Sia $(x_n)$ di Cauchy in $E$. Fissato $\varepsilon>0$, per uniforme continuità esiste $\delta>0$ (dipendente solo da $\varepsilon$) tale che $|x-y|<\delta\Rightarrow|f(x)-f(y)|<\varepsilon$, per ogni $x,y\in E$. Poiché $(x_n)$ è di Cauchy, esiste $N$ con $n,m>N\Rightarrow|x_n-x_m|<\delta$ (usando proprio questo $\delta$, che non dipende dal punto — qui sta il ruolo essenziale dell'**uniformità**: con la sola continuità puntuale, il $\delta$ potrebbe dipendere dal punto specifico $x_n$, rendendo impossibile fissarne uno solo per il criterio di Cauchy). Per $n,m>N$: $|x_n-x_m|<\delta\Rightarrow|f(x_n)-f(x_m)|<\varepsilon$: $(f(x_n))$ è di Cauchy.

## Riepilogo finale

- La **continuità uniforme** richiede un $\delta$ che dipenda solo da $\varepsilon$, non dal punto: nella scrittura dei quantificatori, $\forall\varepsilon\,\exists\delta\,\forall x_0$ invece di $\forall x_0\,\forall\varepsilon\,\exists\delta$.
- $1/x$ su $(0,1)$ è continua ma non uniformemente continua: il $\delta$ necessario si restringe senza limite avvicinandosi a $0$, che non appartiene all'insieme.
- Il **teorema di Heine-Cantor**: continua su un **compatto** implica automaticamente uniformemente continua — la compattezza elimina ogni "via di fuga" verso un punto mancante.
- La continuità uniforme sarà l'ingrediente essenziale per dimostrare, nel notebook 10.15, che ogni funzione continua su $[a,b]$ è integrabile secondo Riemann.
