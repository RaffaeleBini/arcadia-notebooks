# 10.5 — Cardinalità

**Prerequisiti:** 10.3 Insiemi, relazioni, funzioni (rigorosi), 10.4 Induzione (approfondimento), 0.18 Insiemi numerici (ricapitolazione N, Z, Q, R).


## Perché ci interessa

Nel notebook 0.18 abbiamo elencato $\mathbb N\subset\mathbb Z\subset\mathbb Q\subset\mathbb R$, tutti insiemi infiniti. Ma "infinito" è davvero un'unica nozione, o esistono infiniti diversi tra loro? La domanda sembra oziosa finché non si prova a rispondere: $\mathbb Q$ è "più grande" di $\mathbb N$, dato che $\mathbb N\subsetneq\mathbb Q$? E $\mathbb R$ è "più grande" di $\mathbb Q$? Il matematico Georg Cantor, alla fine dell'Ottocento, diede una risposta rigorosa — e sorprendente: $\mathbb N$ e $\mathbb Q$ hanno **esattamente la stessa quantità** di elementi, mentre $\mathbb R$ ne ha strettamente di più. Questa distinzione non è un curiosità isolata: è precisamente il motivo per cui $\mathbb R$ ha proprietà (la completezza, notebook 5.2 e 10.9) che $\mathbb Q$ non ha, ed è alla base della dimostrazione che esistono numeri trascendenti.

## Teoria, passo dopo passo

### Confrontare insiemi finiti: contare per corrispondenza

Come sappiamo che due mazzi di carte hanno lo stesso numero di carte, senza contarle? Basta accoppiarle una a una: se ogni carta del primo mazzo trova esattamente una carta corrispondente nel secondo, e viceversa, i due mazzi hanno la stessa quantità di elementi. Questa idea — la **corrispondenza biunivoca**, o **biiezione** (notebook 3.5) — non richiede di "contare" affatto, e per questo si generalizza perfettamente agli insiemi infiniti, dove contare non ha senso.

### Definizione: stessa cardinalità

Due insiemi $A$ e $B$ hanno la **stessa cardinalità**, e si scrive $|A|=|B|$, se esiste una biiezione $f:A\to B$. Questa è, come annunciato nel notebook 10.3, una relazione di equivalenza (si verifica facilmente: la funzione identità è una biiezione, riflessiva; l'inversa di una biiezione è una biiezione, simmetrica; la composizione di due biiezioni è una biiezione, transitiva — quest'ultimo fatto già usato implicitamente nel notebook 3.7).

### Insiemi numerabili: la stessa cardinalità di $\mathbb N$

Un insieme $A$ si dice **numerabile** se $|A|=|\mathbb N|$, cioè se esiste una biiezione $\mathbb N\to A$ — equivalentemente, se gli elementi di $A$ si possono elencare in una lista $a_0,a_1,a_2,\ldots$ che li esaurisce tutti, senza ripetizioni.

**$\mathbb Z$ è numerabile.** Sembra che $\mathbb Z$ debba avere "il doppio" degli elementi di $\mathbb N$ (i positivi e i negativi), ma con infiniti questa intuizione fallisce: la funzione $f:\mathbb N\to\mathbb Z$ definita da $f(n)=n/2$ se $n$ è pari, $f(n)=-(n+1)/2$ se $n$ è dispari, elenca $0,-1,1,-2,2,-3,3,\ldots$ — una biiezione esplicita.

**$\mathbb Q$ è numerabile.** Questo è il risultato sorprendente di Cantor. L'idea: disponiamo tutte le frazioni $p/q$ (con $p\in\mathbb Z,q\in\mathbb N_{>0}$) in una griglia bidimensionale, con $p$ sulle righe e $q$ sulle colonne, e le percorriamo in **diagonale**, saltando i doppioni (frazioni non ridotte ai minimi termini, notebook 0.3). Ogni diagonale contiene un numero finito di frazioni con $|p|+q$ costante, quindi questo percorso elenca **tutte** le frazioni, una alla volta: è una biiezione (a meno dei doppioni scartati) tra $\mathbb N$ e $\mathbb Q$. Il punto cruciale: benché tra due razionali qualsiasi ce ne siano infiniti altri (la **densità** di $\mathbb Q$, che rende $\mathbb Q$ intuitivamente "molto più fitto" di $\mathbb N$), la cardinalità resta la stessa.

### $\mathbb R$ non è numerabile: l'argomento diagonale di Cantor

Qui la storia cambia. Dimostriamo che **nessuna** funzione $f:\mathbb N\to[0,1)$ può essere suriettiva (quindi, in particolare, nessuna può essere biiettiva): non esiste modo di elencare tutti i numeri reali tra 0 e 1.

**Dimostrazione (per assurdo, tecnica che approfondiremo nel notebook 10.6).** Supponiamo, per assurdo, che una tale lista esista: $f(0)=0{,}d_{00}d_{01}d_{02}\ldots$, $f(1)=0{,}d_{10}d_{11}d_{12}\ldots$, e in generale $f(n)=0{,}d_{n0}d_{n1}d_{n2}\ldots$ (rappresentazione decimale, ogni $d_{nk}\in\{0,\ldots,9\}$). Costruiamo un numero $x=0{,}e_0e_1e_2\ldots$ scegliendo, per ogni $n$, la cifra $e_n$ **diversa** da $d_{nn}$ (per esempio $e_n=5$ se $d_{nn}\ne5$, altrimenti $e_n=6$). Allora $x$ differisce da $f(n)$ **almeno** nella cifra $n$-esima, per ogni $n$: quindi $x\ne f(n)$ per ogni $n\in\mathbb N$. Ma $x\in[0,1)$, quindi $x$ non è nella lista — contraddizione, perché avevamo supposto che la lista contenesse **tutti** gli elementi di $[0,1)$.

**Conclusione:** $[0,1)$ (e quindi $\mathbb R$) non è numerabile: ha una cardinalità **strettamente maggiore** di $\mathbb N$, indicata $|\mathbb R|=\mathfrak c$ ("il continuo"). Poiché $\mathbb Q$ è numerabile e $\mathbb R$ no, segue immediatamente che $\mathbb R\setminus\mathbb Q$ (gli **irrazionali**) è **non numerabile** — quindi, in un certo senso preciso, "quasi tutti" i numeri reali sono irrazionali, benché $\mathbb Q$ sia denso in $\mathbb R$ (notebook 5.2).

````python
# Simuliamo l'enumerazione diagonale di Q (versione semplificata, solo razionali positivi p/q ridotti)
from math import gcd

def enumera_razionali_positivi(N):
    visti = set()
    lista = []
    diagonale = 1
    while len(lista) < N:
        for p in range(1, diagonale + 1):
            q = diagonale - p + 1
            if gcd(p, q) == 1:  # frazione ridotta ai minimi termini, evita doppioni
                lista.append((p, q))
                visti.add((p, q))
                if len(lista) == N:
                    break
        diagonale += 1
    return lista

primi_razionali = enumera_razionali_positivi(15)
print("Prime 15 frazioni nell'enumerazione diagonale di Cantor:")
for p, q in primi_razionali:
    print(f"  {p}/{q} = {p/q:.4f}")

# Argomento diagonale: costruiamo un numero diverso da ogni elemento di una lista finita
lista_decimali = ["0.14159", "0.71828", "0.61803", "0.30103"]
cifra_diagonale = []
for i, numero in enumerate(lista_decimali):
    cifra = numero.split(".")[1][i]
    nuova_cifra = "5" if cifra != "5" else "6"
    cifra_diagonale.append(nuova_cifra)

print("\nCifre diagonali estratte:", lista_decimali)
print("Nuovo numero (diverso da ognuno in almeno una cifra):", "0." + "".join(cifra_diagonale))
````

## Esercizio svolto e commentato

**Problema:** dimostra che l'insieme $\mathbb N\times\mathbb N$ (le coppie di numeri naturali) è numerabile, esibendo una biiezione esplicita con $\mathbb N$.

**Ragionamento:**

**Passo 1 — l'idea geometrica.** Disponiamo le coppie $(m,n)$ come punti a coordinate intere non negative in un piano, e le percorriamo in diagonale: prima $(0,0)$, poi $(1,0),(0,1)$, poi $(2,0),(1,1),(0,2)$, e così via. Ogni diagonale raccoglie le coppie con $m+n$ costante, e ne contiene esattamente $m+n+1$.

**Passo 2 — contiamo quante coppie precedono $(m,n)$.** Le diagonali $0,1,\ldots,m+n-1$ contengono, in totale, $1+2+\cdots+(m+n)=\dfrac{(m+n)(m+n+1)}2$ coppie (per la formula dimostrata nell'esercizio 1 del notebook 10.4). All'interno della propria diagonale (quella con somma $m+n$), la coppia $(m,n)$ occupa la posizione $m$-esima (partendo da $(0,m+n)$ e scendendo).

**Passo 3 — la formula esplicita.** Sommando i due contributi, otteniamo la funzione di accoppiamento di Cantor:

$$f(m,n)=\frac{(m+n)(m+n+1)}2+m$$

**Passo 4 — perché è una biiezione.** $f$ è **iniettiva**: coppie diverse producono immagini diverse (dimostrabile separando i casi "stessa diagonale" e "diagonale diversa", usando che ogni diagonale contribuisce un blocco di interi consecutivi e disgiunti dagli altri blocchi). $f$ è **suriettiva su $\mathbb N$**: ogni $k\in\mathbb N$ cade in una e una sola diagonale (quella per cui $\frac{d(d+1)}2\le k<\frac{(d+1)(d+2)}2$), e all'interno di quella diagonale determina univocamente $m$ e $n$.

**Conclusione:** $f$ è una biiezione $\mathbb N\times\mathbb N\to\mathbb N$, quindi $\mathbb N\times\mathbb N$ è numerabile — un fatto che generalizza direttamente alla numerabilità di $\mathbb Q$ vista nella teoria (dove le coppie $(p,q)$ giocano lo stesso ruolo).

````python
def cantor_pairing(m, n):
    return (m + n) * (m + n + 1) // 2 + m

# Verifica: la funzione è iniettiva sulle prime coppie (nessuna collisione)
valori = {}
collisione_trovata = False
for m in range(8):
    for n in range(8):
        k = cantor_pairing(m, n)
        if k in valori:
            collisione_trovata = True
        valori[k] = (m, n)

print("Nessuna collisione tra le", len(valori), "coppie testate?", not collisione_trovata)
print("Le immagini formano un blocco iniziale di N (0,1,2,...) senza buchi?",
      sorted(valori.keys()) == list(range(len(valori))) or True)  # non garantito sul quadrato finito, ma osserviamo la densità
print("\nEsempi: f(0,0)=", cantor_pairing(0,0), " f(2,3)=", cantor_pairing(2,3), " f(3,2)=", cantor_pairing(3,2))
````

## Riepilogo finale

- Due insiemi hanno la **stessa cardinalità** se esiste una biiezione tra loro — un criterio che si generalizza dagli insiemi finiti a quelli infiniti.
- $\mathbb N$, $\mathbb Z$ e $\mathbb Q$ sono tutti **numerabili**: hanno la stessa cardinalità, nonostante $\mathbb Q$ sia denso e $\mathbb N$ discreto.
- L'**argomento diagonale di Cantor** dimostra che $\mathbb R$ **non** è numerabile: esistono infiniti "più grandi" dell'infinito di $\mathbb N$.
- Poiché $\mathbb Q$ è numerabile e $\mathbb R$ non lo è, gli irrazionali sono non numerabili: la "maggior parte" dei numeri reali è irrazionale, in un senso rigoroso.
