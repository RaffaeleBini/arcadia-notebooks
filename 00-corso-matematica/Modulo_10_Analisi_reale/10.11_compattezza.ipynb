## Perché ci interessa

Nel notebook 5.7 abbiamo enunciato il teorema di Weierstrass: "una funzione continua su un intervallo chiuso e limitato $[a,b]$ ammette massimo e minimo" — ma perché servono **entrambe** queste ipotesi sull'intervallo? Il notebook 10.7 ha già mostrato un controesempio se l'intervallo è aperto. La proprietà che rende $[a,b]$ "abbastanza buono" da garantire l'esistenza del massimo si chiama **compattezza**, e questo notebook la definisce con precisione, dimostrando che per i sottoinsiemi di $\mathbb R$ (e più in generale di $\mathbb R^n$, ricollegandoci alla topologia del notebook 7.3) essere compatti equivale, in modo sorprendentemente concreto, a essere **chiusi e limitati**. È l'ultimo tassello prima di poter finalmente dimostrare rigorosamente il teorema di Weierstrass, nel prossimo notebook.

## Teoria, passo dopo passo

### Ricoprimenti aperti: la definizione topologica

Un **ricoprimento aperto** di un insieme $K\subseteq\mathbb R$ è una famiglia di insiemi aperti $\{U_i\}_{i\in I}$ (nel senso del notebook 7.3: unioni di intervalli aperti) tale che $K\subseteq\bigcup_{i\in I}U_i$ — ogni punto di $K$ appartiene ad almeno un $U_i$. La famiglia può essere **infinita**, persino non numerabile.

**Definizione (compattezza).** $K$ è **compatto** se **ogni** ricoprimento aperto di $K$ ammette un **sottoricoprimento finito**: cioè, da qualunque famiglia (anche infinita) di aperti che ricopre $K$, si possono sempre selezionare un numero **finito** di essi che già ricoprono tutto $K$.

Questa definizione, all'apparenza astratta, cattura un'idea molto concreta: $K$ è "abbastanza piccolo" da non poter "sfuggire" a un ricoprimento finito. Vediamo perché un intervallo aperto **non** è compatto, con un controesempio esplicito (tecnica del notebook 10.7): il ricoprimento $U_n=\left(\dfrac1n,1\right)$ per $n=2,3,4,\ldots$ ricopre $(0,1)$ (ogni $x\in(0,1)$ sta in $U_n$ per $n$ abbastanza grande, cioè $n>1/x$, per Archimede), ma **nessuna** sottofamiglia finita $U_{n_1},\ldots,U_{n_k}$ può ricoprire $(0,1)$: prendendo $N=\max(n_1,\ldots,n_k)$, il punto $\dfrac1{2N}$ resta scoperto da tutti gli $U_{n_j}$ (perché $\dfrac1{2N}<\dfrac1{n_j}$ per ogni $j$). Nessun sottoricoprimento finito esiste: $(0,1)$ non è compatto.

### Teorema di Heine-Borel: compatto ⟺ chiuso e limitato

**Teorema (Heine-Borel, in $\mathbb R$).** $K\subseteq\mathbb R$ è compatto $\iff$ $K$ è chiuso e limitato.

**Dimostrazione ($\Rightarrow$, compatto $\Rightarrow$ limitato).** Il ricoprimento $\{(-n,n):n\in\mathbb N\}$ copre tutto $\mathbb R$, quindi copre $K$. Per compattezza, esiste un sottoricoprimento finito $(-n_1,n_1),\ldots,(-n_k,n_k)$; ponendo $N=\max(n_1,\ldots,n_k)$, si ha $K\subseteq(-N,N)$: $K$ è limitato.

**Dimostrazione ($\Rightarrow$, compatto $\Rightarrow$ chiuso).** Si dimostra che il complementare $K^c$ è aperto (definizione di chiuso, notebook 7.3): dato $x\notin K$, per ogni $y\in K$ scegliamo un intorno $U_y$ di $y$ e un intorno $V_y$ di $x$ disgiunti (possibile perché $x\ne y$, prendendo raggi minori di $|x-y|/2$). Gli $U_y$ (al variare di $y\in K$) formano un ricoprimento aperto di $K$; per compattezza, un numero finito $U_{y_1},\ldots,U_{y_k}$ basta a ricoprire $K$. L'intersezione finita $V=V_{y_1}\cap\cdots\cap V_{y_k}$ è un intorno aperto di $x$ (intersezione finita di aperti è aperta, notebook 7.3) disgiunto da ogni $U_{y_i}$, quindi disgiunto da $K$: $V\subseteq K^c$. Ogni punto di $K^c$ ha un intorno contenuto in $K^c$: $K^c$ è aperto, cioè $K$ è chiuso.

**Dimostrazione ($\Leftarrow$, chiuso e limitato $\Rightarrow$ compatto), traccia.** Basta dimostrarlo per $K=[a,b]$ (il caso generale segue perché un chiuso contenuto in un intervallo compatto eredita la compattezza, un fatto che si dimostra con un argomento analogo a quello sopra). Si procede per assurdo con la bisezione, esattamente come nella dimostrazione di Bolzano-Weierstrass (notebook 10.10): se un ricoprimento aperto di $[a,b]$ non avesse sottoricoprimenti finiti, dimezzando ripetutamente si otterrebbero intervalli incapsulati, nessuno dei quali ricopribile finitamente, che per il teorema degli intervalli incapsulati (notebook 10.9) si restringono a un punto $p\in[a,b]$; ma $p$ appartiene a qualche aperto $U$ del ricoprimento, e $U$ conterrebbe interamente uno degli intervalli piccoli (essendo aperto), fornendo un sottoricoprimento finito (fatto da un solo insieme) — contraddizione.

### Compattezza sequenziale: l'equivalenza con Bolzano-Weierstrass

**Teorema.** $K\subseteq\mathbb R$ è compatto $\iff$ ogni successione in $K$ ha una sottosuccessione convergente **a un punto di $K$** (compattezza sequenziale).

Questa equivalenza collega direttamente la definizione topologica (ricoprimenti) a Bolzano-Weierstrass (notebook 10.10): se $K$ è chiuso e limitato, ogni successione in $K$ è limitata, quindi per Bolzano-Weierstrass ha una sottosuccessione convergente; il limite appartiene a $K$ perché $K$ è chiuso (un insieme chiuso contiene tutti i limiti delle proprie successioni convergenti, per definizione di chiusura, notebook 7.3). È questa versione **sequenziale** della compattezza — più intuitiva del linguaggio dei ricoprimenti — quella che useremo davvero nelle dimostrazioni dei prossimi notebook.

````python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 3))

# Controesempio: (0,1) non compatto -- il ricoprimento U_n = (1/n, 1) non ha sottoricoprimento finito
n_valori = [2, 3, 5, 10, 30]
for i, n in enumerate(n_valori):
    ax.plot([1/n, 1], [i, i], linewidth=4, solid_capstyle="butt", label=f"U_{n} = (1/{n}, 1)")

ax.set_yticks(range(len(n_valori)))
ax.set_yticklabels([f"U_{n}" for n in n_valori])
ax.set_xlim(-0.05, 1.05)
ax.set_title("Nessuna unione finita di questi U_n copre (0,1): il punto 1/(2N) resta sempre scoperto")
ax.set_xlabel("x")
plt.tight_layout()
plt.show()

# Verifica numerica: per ogni scelta finita di indici, esiste un punto scoperto
indici_scelti = [2, 5, 10]
N = max(indici_scelti)
punto_scoperto = 1 / (2 * N)
print(f"Con gli indici {indici_scelti}, il punto x = 1/(2*{N}) = {punto_scoperto:.4f} è scoperto,")
print(f"perché è < 1/n per ogni n in {indici_scelti}:", all(punto_scoperto < 1/n for n in indici_scelti))
````

## Esercizio svolto e commentato

**Problema:** dimostra che $K=\{0\}\cup\left\{\dfrac1n:n\in\mathbb N_{>0}\right\}$ è compatto, esibendo direttamente, per un ricoprimento aperto arbitrario, un sottoricoprimento finito.

**Ragionamento:**

**Passo 1 — osserviamo la struttura di $K$.** $K=\{0,1,1/2,1/3,\ldots\}$: è chiuso (contiene il proprio punto di accumulazione $0$, verso cui $1/n\to0$) e limitato (contenuto in $[0,1]$). Per Heine-Borel, ci aspettiamo che sia compatto — lo dimostriamo direttamente dalla definizione, per esercitare il ragionamento sui ricoprimenti.

**Passo 2 — sia $\{U_i\}$ un ricoprimento aperto arbitrario di $K$.** Poiché $0\in K$, esiste un indice $i_0$ con $0\in U_{i_0}$. Essendo $U_{i_0}$ aperto, contiene un intero intervallo $(-\delta,\delta)$ per qualche $\delta>0$.

**Passo 3 — usiamo la proprietà archimedea (notebook 10.9).** Esiste $N\in\mathbb N$ con $\dfrac1N<\delta$. Quindi **tutti** i punti $1/n$ con $n\ge N$ soddisfano $1/n<\delta$, cioè appartengono a $(-\delta,\delta)\subseteq U_{i_0}$: sono già coperti da un **solo** insieme del ricoprimento.

**Passo 4 — restano scoperti solo i finiti punti $1,1/2,\ldots,1/(N-1)$.** Ciascuno appartiene a $K$, quindi a qualche $U_{i_1},\ldots,U_{i_{N-1}}$ del ricoprimento originale (per definizione di ricoprimento).

**Conclusione:** $U_{i_0},U_{i_1},\ldots,U_{i_{N-1}}$ è un sottoricoprimento **finito** di $K$, estratto da un ricoprimento arbitrario: $K$ è compatto. **L'idea chiave, generale**: il punto di accumulazione $0$ "assorbe" con un solo aperto **infiniti** punti della successione, lasciando solo un residuo finito da coprire separatamente — questo è esattamente il meccanismo che rende compatti gli insiemi chiusi e limitati, e fallisce se il punto di accumulazione (qui $0$) viene rimosso dall'insieme, come nel controesempio $(0,1)$ della teoria.

````python
# Verifica numerica dell'idea chiave: per ogni delta, un numero finito di punti 1/n sta fuori da (-delta, delta)
def punti_fuori_intorno(delta, n_max=1000):
    return [n for n in range(1, n_max) if 1/n >= delta]

for delta in [0.5, 0.1, 0.01, 0.001]:
    fuori = punti_fuori_intorno(delta)
    print(f"delta={delta}: punti 1/n con n=1..{len(fuori)} restano fuori dall'intorno di 0 "
          f"({len(fuori)} punti, un numero FINITO)")
````

**1.** Siano $K_1,\ldots,K_m$ compatti, e $\{U_i\}$ un ricoprimento aperto di $K_1\cup\cdots\cup K_m$. **2.** Sia $K$ compatto e $C$ chiuso, e sia $\{U_i\}$ un ricoprimento aperto di $K\cap C$. Aggiungiamo l'aperto $C^c$ (complementare di un chiuso, quindi aperto) alla famiglia: $\{U_i\}\cup\{C^c\}$ ricopre tutto $K$ (ogni punto di $K$ o è in $C$, quindi in $K\cap C$, quindi coperto da qualche $U_i$; oppure non è in $C$, quindi in $C^c$). Per compattezza di $K$, esiste un sottoricoprimento finito, che possiamo assumere includa $C^c$ (se non serve, lo si scarta). Gli insiemi $U_i$ selezionati (escludendo eventualmente $C^c$, che non serve a coprire punti di $C$) formano un sottoricoprimento finito di $K\cap C$.

**3.** $[0,1]\cup[2,3]$: chiuso (unione finita di chiusi) e limitato $\Rightarrow$ **compatto**. $\mathbb Q\cap[0,1]$: limitato ma **non chiuso** (per esempio, contiene una successione di razionali che approssima $\sqrt2/\sqrt2\cdot$... più semplicemente, contiene razionali che approssimano un irrazionale come $1/\sqrt2\in[0,1]$, che non appartiene all'insieme) $\Rightarrow$ **non compatto**. $\{1/n:n\ge1\}$ senza lo $0$: limitato ma **non chiuso** (manca il punto di accumulazione $0$, verso cui converge, a differenza dell'esercizio svolto che includeva $0$) $\Rightarrow$ **non compatto** — coerente con il fatto che l'esercizio svolto funzionava proprio perché $0\in K$.

**4.** Sia $(x_n)$ una successione di punti **distinti** in $F$ (esiste perché $F$ è infinito). Per compattezza sequenziale di $K$ (equivalente a Heine-Borel per il teorema della teoria), $(x_n)$ ha una sottosuccessione $x_{n_k}\to p\in K$. Dimostriamo che $p$ è un punto di accumulazione di $F$: fissato $\delta>0$, per la convergenza esiste $k$ con $x_{n_k}\ne p$ (possibile perché i punti sono distinti, quindi al più uno di essi coincide con $p$) e $|x_{n_k}-p|<\delta$, cioè un punto di $F$ diverso da $p$ nell'intorno di raggio $\delta$: per definizione, $p$ è un punto di accumulazione di $F$, e $p\in K$.

## Riepilogo finale

- $K$ è **compatto** se ogni ricoprimento aperto ammette un sottoricoprimento finito — una definizione topologica astratta ma potente.
- Il **teorema di Heine-Borel** (in $\mathbb R$): $K$ è compatto $\iff$ $K$ è chiuso e limitato, una caratterizzazione molto più concreta.
- La **compattezza sequenziale** (ogni successione in $K$ ha una sottosuccessione convergente in $K$) è equivalente, e collega la compattezza direttamente al teorema di Bolzano-Weierstrass del notebook 10.10.
- La compattezza è la proprietà mancante che spiegava il controesempio del notebook 10.7 ($f(x)=x$ su $(0,1)$ senza massimo): $(0,1)$ non è compatto perché non è chiuso.
