**Prerequisiti:** 0.1 Numeri naturali, 0.20 Introduzione alle dimostrazioni, 10.1 Logica proposizionale (approfondimento).


## Perché ci interessa

Nel corso abbiamo più volte enunciato fatti "per ogni $n$" senza dimostrarli con cura: la formula della somma geometrica $1+q+\cdots+q^n=\frac{1-q^{n+1}}{1-q}$ (usata nel notebook 4.8), la regola $\left(\prod a_i\right)'=\ldots$ per il prodotto di $n$ funzioni, o semplicemente il fatto che ogni numero naturale ha un successore (notebook 0.1). Come si dimostra un'affermazione che riguarda **infiniti** casi, uno per ogni $n\in\mathbb N$? Non possiamo verificarli uno per uno. Il **principio di induzione** risolve esattamente questo problema, ed è l'unico strumento dimostrativo capace di gestire l'infinito dei numeri naturali con un numero *finito* di passaggi logici. Lo riprendiamo ora con tutto il rigore che il modulo richiede, perché lo useremo ripetutamente: per dimostrare la disuguaglianza di Bernoulli, proprietà delle successioni, e persino il fatto (notebook 10.5) che $\mathbb N$ e $\mathbb Q$ hanno "la stessa quantità" di elementi.

## Teoria, passo dopo passo

### L'idea: la fila di tessere del domino

Immagina una fila infinita di tessere del domino, disposte in modo che, se una cade, fa cadere anche la successiva. Se sappiamo che **la prima tessera cade**, e che **ogni tessera che cade fa cadere la successiva**, possiamo concludere che **tutte** le tessere cadono — nessuna esclusa, per quanto lontana nella fila. Questa è, in sostanza, la logica dell'induzione.

### Il principio di induzione (forma debole)

Sia $P(n)$ una proposizione che dipende da $n\in\mathbb N$. Se valgono:

1. **Base**: $P(0)$ è vera (oppure $P(n_0)$, se si parte da un $n_0$ diverso da $0$);
2. **Passo induttivo**: per ogni $n\ge n_0$, $P(n)\Rightarrow P(n+1)$ (assumendo $P(n)$ vera — l'**ipotesi induttiva** — si dimostra che anche $P(n+1)$ è vera);

allora $P(n)$ è vera per **ogni** $n\ge n_0$.

**Perché questo principio è valido**: è, in realtà, uno degli **assiomi** che definiscono i numeri naturali (uno dei cosiddetti assiomi di Peano, che fondano $\mathbb N$ nel notebook 0.1) — non lo si dimostra da fatti più elementari, ma lo si assume come proprietà caratterizzante di $\mathbb N$: è ciò che distingue $\mathbb N$ da un insieme "generico". Intuitivamente: se $P$ valesse per la base ma fallisse da qualche parte, ci sarebbe un **primo** $n$ per cui $P(n)$ è falsa (perché $\mathbb N$ è bene ordinato: ogni sottoinsieme non vuoto ha un minimo, una proprietà che sfrutteremo tra poco nelle dimostrazioni per assurdo); ma allora $P(n-1)$ sarebbe vera (per minimalità di $n$) e il passo induttivo garantirebbe $P(n)$ vera, contraddizione.

### Il principio di induzione forte

Una variante spesso più comoda: nel passo induttivo, invece di assumere solo $P(n)$, si può assumere **tutte** le proposizioni precedenti $P(n_0), P(n_0+1),\ldots,P(n)$ per dedurre $P(n+1)$. È logicamente equivalente alla forma debole, ma a volte semplifica la dimostrazione (per esempio nel teorema fondamentale dell'aritmetica, o per successioni definite da relazioni di ricorrenza che coinvolgono più termini precedenti, come $a_n=a_{n-1}+a_{n-2}$).

````python
import sympy as sp

n, q = sp.symbols('n q')

# Verifica simbolica della somma geometrica per alcuni valori di n:
# 1 + q + q^2 + ... + q^n = (1 - q^(n+1)) / (1 - q)
for valore_n in range(6):
    somma_diretta = sum(q**k for k in range(valore_n + 1))
    formula_chiusa = (1 - q**(valore_n + 1)) / (1 - q)
    differenza = sp.simplify(somma_diretta - formula_chiusa)
    print(f"n={valore_n}: somma - formula chiusa =", differenza)

print("\nLa differenza è sempre 0: la formula regge per ogni n testato (ma solo l'induzione lo dimostra per TUTTI gli n).")
````

## Esercizio svolto e commentato

**Problema:** dimostra per induzione la **disuguaglianza di Bernoulli**: per ogni $x\ge-1$ e per ogni $n\in\mathbb N$,

$$(1+x)^n\ge1+nx$$

**Ragionamento:**

**Base ($n=0$):** $(1+x)^0=1$, e $1+0\cdot x=1$. Vale $1\ge1$: vera.

**Passo induttivo:** assumiamo vera l'**ipotesi induttiva** $(1+x)^n\ge1+nx$ per un $n$ fissato, e dimostriamo che vale per $n+1$, cioè $(1+x)^{n+1}\ge1+(n+1)x$.

Partiamo da $(1+x)^{n+1}=(1+x)^n\cdot(1+x)$. Poiché $x\ge-1$, si ha $1+x\ge0$, quindi moltiplicare per $(1+x)$ **non inverte** le disuguaglianze (una proprietà d'ordine vista nel notebook 5.1, che vale solo quando si moltiplica per quantità non negative). Applichiamo l'ipotesi induttiva, moltiplicando entrambi i membri per $(1+x)\ge0$:

$$(1+x)^n\cdot(1+x)\ge(1+nx)(1+x)$$

Ora sviluppiamo il membro destro: $(1+nx)(1+x)=1+x+nx+nx^2=1+(n+1)x+nx^2$ (prodotto notevole, notebook 1.2). Poiché $nx^2\ge0$ sempre (un numero naturale per un quadrato), possiamo scartare questo termine mantenendo la disuguaglianza:

$$1+(n+1)x+nx^2 \ge 1+(n+1)x$$

Combinando le due disuguaglianze in catena: $(1+x)^{n+1}\ge1+(n+1)x+nx^2\ge1+(n+1)x$, cioè esattamente la tesi per $n+1$.

**Conclusione:** per il principio di induzione, $(1+x)^n\ge1+nx$ vale per ogni $n\in\mathbb N$ e ogni $x\ge-1$. **Perché serve l'ipotesi $x\ge-1$**: è esattamente ciò che garantisce $1+x\ge0$, il passaggio cruciale che permette di moltiplicare senza invertire la disuguaglianza — senza questa ipotesi, il passo induttivo non sarebbe valido.

````python
import sympy as sp

x = sp.symbols('x')

def verifica_bernoulli(x_val, n_max=8):
    for n in range(n_max + 1):
        sinistra = (1 + x_val)**n
        destra = 1 + n * x_val
        print(f"n={n}: (1+x)^n = {float(sinistra):.4f}, 1+nx = {float(destra):.4f}, "
              f"disuguaglianza rispettata? {sinistra >= destra}")

print("Test con x = 0.5 (>= -1):")
verifica_bernoulli(0.5)
print("\nTest con x = -0.5 (>= -1):")
verifica_bernoulli(-0.5)
````

3. Dimostra per induzione che $\displaystyle\sum_{k=0}^n q^k=\frac{1-q^{n+1}}{1-q}$ per ogni $n\ge0$ e ogni $q\ne1$ (la formula usata senza dimostrazione nel notebook 4.8).
4. *(più impegnativo)* La successione di Fibonacci è definita da $F_0=0,F_1=1,F_n=F_{n-1}+F_{n-2}$. Usa il **principio di induzione forte** per dimostrare che $F_n<2^n$ per ogni $n\ge0$ (suggerimento: nel passo induttivo, per dedurre il caso $n$ avrai bisogno sia di $F_{n-1}$ sia di $F_{n-2}$, cioè di due casi precedenti, non solo dell'ultimo).

**1. Base** ($n=1$): $\sum_{k=1}^1 k=1$, e $\frac{1\cdot2}2=1$. Vero. **Passo**: assumendo $\sum_{k=1}^n k=\frac{n(n+1)}2$, calcoliamo $\sum_{k=1}^{n+1}k=\sum_{k=1}^nk+(n+1)=\frac{n(n+1)}2+(n+1)=(n+1)\left(\frac n2+1\right)=(n+1)\cdot\frac{n+2}2=\frac{(n+1)(n+2)}2$, che è la formula per $n+1$. Per induzione, vale per ogni $n\ge1$.

**2. Base** ($n=1$): $1!=1\ge2^0=1$. Vero. **Passo**: assumendo $n!\ge2^{n-1}$, calcoliamo $(n+1)!=(n+1)\cdot n!\ge(n+1)\cdot2^{n-1}$ (moltiplicando l'ipotesi induttiva per $n+1>0$, che non inverte la disuguaglianza). Poiché $n\ge1$ implica $n+1\ge2$, si ha $(n+1)\cdot2^{n-1}\ge2\cdot2^{n-1}=2^n$. Combinando: $(n+1)!\ge2^n$, la tesi per $n+1$.

**3. Base** ($n=0$): $\sum_{k=0}^0q^k=1$, e $\frac{1-q}{1-q}=1$ (per $q\ne1$). Vero. **Passo**: assumendo $\sum_{k=0}^nq^k=\frac{1-q^{n+1}}{1-q}$, calcoliamo $\sum_{k=0}^{n+1}q^k=\sum_{k=0}^nq^k+q^{n+1}=\frac{1-q^{n+1}}{1-q}+q^{n+1}=\frac{1-q^{n+1}+q^{n+1}(1-q)}{1-q}=\frac{1-q^{n+1}+q^{n+1}-q^{n+2}}{1-q}=\frac{1-q^{n+2}}{1-q}$, la formula per $n+1$.

**4. Base** ($n=0,1$): $F_0=0<2^0=1$, $F_1=1<2^1=2$. Entrambe vere (verifichiamo due casi base perché il passo userà due termini precedenti). **Passo** (per $n\ge2$, assumendo l'ipotesi forte $F_j<2^j$ per **tutti** i $j<n$): $F_n=F_{n-1}+F_{n-2}<2^{n-1}+2^{n-2}$ (per ipotesi induttiva su entrambi i termini) $=2^{n-2}(2+1)=3\cdot2^{n-2}<4\cdot2^{n-2}=2^n$ (usando $3<4$). Quindi $F_n<2^n$, la tesi. Per induzione forte, vale per ogni $n\ge0$.

Verifica computazionale:
```python
def fib(n):
    F = [0, 1]
    for i in range(2, n + 1):
        F.append(F[i-1] + F[i-2])
    return F[n]

print("Verifica F_n < 2^n per n=0..15:")
for n in range(16):
    print(n, fib(n), 2**n, fib(n) < 2**n)
```

## Riepilogo finale

- Il **principio di induzione** dimostra un'affermazione $P(n)$ per tutti i $n\in\mathbb N$ in due soli passi: la **base** ($P(n_0)$ vera) e il **passo induttivo** ($P(n)\Rightarrow P(n+1)$).
- È un assioma dei numeri naturali (uno degli assiomi di Peano), non un teorema derivato da qualcos'altro: caratterizza $\mathbb N$ come l'insieme "raggiungibile partendo da $0$ e sommando $1$ ripetutamente".
- La forma **forte** permette di usare, nel passo induttivo, tutte le proposizioni precedenti (non solo l'ultima) — utile per successioni ricorsive come Fibonacci.
- L'induzione sarà lo strumento portante di molte dimostrazioni del resto del modulo, incluso il confronto tra $\mathbb N$ e $\mathbb Q$ nel prossimo notebook.
