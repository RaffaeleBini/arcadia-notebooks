# 10.7 — Controesempi

**Prerequisiti:** 10.2 Quantificatori (approfondimento), 10.6 Dimostrazioni dirette, per contrapposizione, per assurdo (approfondimento).


## Perché ci interessa

Fino a ora abbiamo sempre dimostrato che qualcosa **è vero**. Ma metà del lavoro di un matematico consiste nel capire cosa **non** è vero: una congettura plausibile, un'estensione naturale di un teorema, un'inversione di un'implicazione — molte di queste affermazioni sono false, e riconoscerlo con un singolo **controesempio** è spesso più rapido (e altrettanto rigoroso) di una lunga dimostrazione. Nei prossimi notebook incontreremo teoremi con ipotesi delicate — "$f$ continua **su un compatto**" (notebook 10.12), "convergenza **uniforme**" (notebook 10.16) — e capire perché ogni ipotesi è necessaria richiede sempre di sapere costruire un controesempio quando l'ipotesi viene rimossa.

## Teoria, passo dopo passo

### La logica del controesempio

Nel notebook 10.2 abbiamo visto che negare $\forall x\,P(x)$ dà $\exists x\,\neg P(x)$: per dimostrare che "per ogni $x$ vale $P(x)$" è **falsa**, basta esibire **un solo** $x$ per cui $P(x)$ è falsa. Questo $x$ è un **controesempio**. Non serve capire perché la proposizione fallisce "in generale": basta un caso concreto, verificabile direttamente. Questo è ciò che distingue un controesempio da una dimostrazione: mentre dimostrare "per ogni $x$, $P(x)$" richiede un argomento che copra **tutti** i casi, confutarla richiede solo **un** caso.

### Un errore comune: confondere "non dimostrato" con "falso"

Non trovare subito un controesempio non dimostra che la proposizione sia vera: significa solo che non l'abbiamo (ancora) trovato, oppure che la proposizione è vera e va dimostrata. Il valore del controesempio sta proprio nella sua natura definitiva: se lo si trova, la questione è chiusa (la proposizione è falsa); se non lo si trova, resta aperta.

### Controesempi a inversioni di teoremi

Una fonte classica di controesempi: prendere un teorema vero $p\Rightarrow q$ e chiedersi se vale anche il converso $q\Rightarrow p$ (distinzione introdotta nel notebook 10.1). Spesso no. Esempio: nel notebook 5.8 abbiamo visto che **derivabile $\Rightarrow$ continua**. Vale il converso, "continua $\Rightarrow$ derivabile"? La funzione $f(x)=|x|$ è continua in $x=0$ (il limite da sinistra e da destra coincidono e valgono $f(0)=0$) ma non derivabile in $x=0$ (il rapporto incrementale ha limite destro $+1$ e limite sinistro $-1$, diversi — notebook 5.8). Questo singolo esempio è sufficiente per concludere che il converso è falso: **continuità non implica derivabilità**.

### Controesempi a "ipotesi dimenticate"

Un'altra fonte: prendere un teorema vero con ipotesi $H_1\wedge H_2\Rightarrow q$, e verificare se basta $H_1$ da solo. Vedremo, per esempio, che "$f$ continua su un **intervallo chiuso e limitato** $[a,b]$ $\Rightarrow$ $f$ ammette massimo" (teorema di Weierstrass, notebook 5.7, che dimostreremo rigorosamente nel notebook 10.12 via compattezza). Se rimuoviamo l'ipotesi "chiuso e limitato", il teorema fallisce: $f(x)=x$ su $(0,1)$ (intervallo aperto) è continua ma non ammette massimo (l'estremo superiore è 1, ma non è raggiunto da nessun punto dell'intervallo aperto).

````python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# Controesempio 1: continua ma non derivabile (|x| in 0)
x = np.linspace(-2, 2, 400)
axes[0].plot(x, np.abs(x), color="steelblue")
axes[0].set_title("f(x) = |x|: continua ma non derivabile in 0")
axes[0].axvline(0, color="gray", linewidth=0.5)
axes[0].axhline(0, color="gray", linewidth=0.5)

# Controesempio 2: continua su intervallo aperto, senza massimo
x2 = np.linspace(0.001, 0.999, 400)
axes[1].plot(x2, x2, color="firebrick")
axes[1].plot(1, 1, marker="o", markerfacecolor="none", markeredgecolor="firebrick",
             label="estremo superiore non raggiunto")
axes[1].set_title("f(x) = x su (0,1): continua ma senza massimo")
axes[1].legend()

plt.tight_layout()
plt.show()

print("f(x)=|x| non è derivabile in 0: rapporto incrementale destro e sinistro diversi.")
print("Limite destro del rapporto incrementale:", 1)
print("Limite sinistro del rapporto incrementale:", -1)
````

## Esercizio svolto e commentato

**Problema:** è vero che "se una successione $(a_n)$ è limitata, allora converge"? Se falso, esibisci un controesempio.

**Ragionamento:**

**Passo 1 — cerchiamo di ricordare un teorema simile.** Nel notebook 4.4 abbiamo incontrato, in modo informale, "successione **monotona e** limitata $\Rightarrow$ convergente" (teorema di monotonia, che dimostreremo rigorosamente nel notebook 10.10). La proposizione da esaminare rimuove l'ipotesi di monotonia: dobbiamo chiederci se resta vera.

**Passo 2 — proviamo a costruire un controesempio.** Cerchiamo una successione limitata che **non** si stabilizzi verso un valore. L'idea più semplice: una successione che oscilla avanti e indietro tra due valori, senza mai avvicinarsi definitivamente a nessuno dei due.

**Passo 3 — verifichiamo il candidato $a_n=(-1)^n$.** È limitata: $|a_n|=1$ per ogni $n$, quindi $M=1$ funziona nella definizione di limitatezza (notebook 4.5). Non converge: i termini pari valgono sempre $1$, i termini dispari sempre $-1$; se convergesse a un limite $L$, ogni sottosuccessione dovrebbe convergere allo stesso $L$ (un fatto che dimostreremo rigorosamente nel notebook 10.10), ma la sottosuccessione dei pari tende a $1$ e quella dei dispari a $-1$, che sono diversi — quindi $(-1)^n$ non converge.

**Conclusione:** $a_n=(-1)^n$ è un controesempio: limitata ma non convergente. La proposizione è **falsa**. **Il ruolo mancante della monotonia:** questo controesempio chiarisce esattamente perché il teorema del notebook 4.4 richiedeva **entrambe** le ipotesi — la monotonia impedisce proprio questo tipo di oscillazione perenne, "incanalando" la successione verso un unico valore limite.

````python
import numpy as np

n = np.arange(0, 20)
a = (-1.0)**n

print("Primi 20 termini di a_n = (-1)^n:", a.astype(int))
print("Successione limitata? |a_n| <= 1 per ogni n:", np.all(np.abs(a) <= 1))
print("Sottosuccessione pari (n=0,2,4,...):", a[::2][:5], "-> tende a 1")
print("Sottosuccessione dispari (n=1,3,5,...):", a[1::2][:5], "-> tende a -1")
print("Limiti diversi per due sottosuccessioni => la successione non converge.")
````

## Riepilogo finale

- Un **controesempio** confuta $\forall x\,P(x)$ esibendo un solo $x$ con $P(x)$ falsa: è la traduzione operativa della regola di negazione $\neg(\forall x\,P)\Leftrightarrow\exists x\,\neg P$ (notebook 10.2).
- Non trovare un controesempio non dimostra che una proposizione sia vera.
- I controesempi più istruttivi nascono da due fonti: **invertire** un'implicazione vera (continua $\ne$ derivabile), o **indebolire** le ipotesi di un teorema vero (compattezza dell'intervallo nel teorema di Weierstrass).
- Sapere costruire controesempi è essenziale per capire, nei prossimi notebook, **perché** ogni ipotesi di un teorema è davvero necessaria.
