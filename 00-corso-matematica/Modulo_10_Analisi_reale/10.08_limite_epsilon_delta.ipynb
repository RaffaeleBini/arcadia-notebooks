# 10.8 — Limite ε-δ

**Prerequisiti:** 5.4 Limiti, 10.2 Quantificatori (approfondimento), 10.7 Controesempi.


## Perché ci interessa

Nel notebook 5.4 abbiamo definito il limite in modo intuitivo: "$f(x)$ tende a $L$ quando $x$ tende a $x_0$" se $f(x)$ "si avvicina quanto vogliamo" a $L$ man mano che $x$ "si avvicina" a $x_0$. Funziona per calcolare, ma le parole "quanto vogliamo" e "si avvicina" non sono formule: non possiamo costruirci sopra una dimostrazione rigorosa. Questo notebook traduce quell'intuizione in una definizione **precisa**, fatta di quantificatori (notebook 10.2) applicati a disuguaglianze concrete — la definizione **ε-δ**, inventata da Cauchy e Weierstrass nell'Ottocento proprio per eliminare ogni ambiguità dal concetto di limite. È la definizione su cui poggeranno tutte le dimostrazioni rigorose del resto del modulo: continuità (10.12), derivabilità (10.14), integrabilità (10.15).

## Teoria, passo dopo passo

### Tradurre "si avvicina quanto vogliamo" in formule

"$f(x)$ è vicino a $L$ quanto vogliamo" significa: per ogni margine di tolleranza $\varepsilon>0$ (per quanto piccolo), possiamo garantire $|f(x)-L|<\varepsilon$ — cioè $f(x)$ dista da $L$ meno di $\varepsilon$. "Purché $x$ sia abbastanza vicino a $x_0$" significa: esiste una distanza $\delta>0$ tale che, se $|x-x_0|<\delta$ (e $x\ne x_0$, perché non ci interessa il valore di $f$ *in* $x_0$, solo il suo comportamento nell'intorno), allora la garanzia sopra vale.

### La definizione ε-δ

$$\lim_{x\to x_0}f(x)=L \quad\stackrel{\text{def}}{\Longleftrightarrow}\quad \forall\varepsilon>0\,\exists\delta>0:\; 0<|x-x_0|<\delta\Rightarrow|f(x)-L|<\varepsilon$$

Il gioco logico da capire: **l'avversario sceglie $\varepsilon$ (la tolleranza, per quanto piccola), noi rispondiamo con un $\delta$** (che quindi **può dipendere da $\varepsilon$** — esattamente come nel notebook 10.2, dove il testimone $y$ poteva dipendere dalla variabile $x$ quantificata prima). Se, qualunque $\varepsilon$ l'avversario scelga, riusciamo sempre a trovare un $\delta$ che funziona, il limite vale $L$.

### Un esempio dimostrato passo passo: $\lim_{x\to2}(3x-1)=5$

**Obiettivo**: dato $\varepsilon>0$ arbitrario, trovare $\delta>0$ tale che $0<|x-2|<\delta\Rightarrow|(3x-1)-5|<\varepsilon$.

**Lavoro preparatorio** (non fa parte della dimostrazione finale, ma guida la scelta di $\delta$): $|(3x-1)-5|=|3x-6|=3|x-2|$. Vogliamo $3|x-2|<\varepsilon$, cioè $|x-2|<\varepsilon/3$. Questo suggerisce di scegliere $\delta=\varepsilon/3$.

**Dimostrazione formale**: dato $\varepsilon>0$, scegliamo $\delta=\varepsilon/3$. Se $0<|x-2|<\delta$, allora $|(3x-1)-5|=3|x-2|<3\delta=3\cdot\dfrac\varepsilon3=\varepsilon$. Quindi la condizione è verificata per ogni $\varepsilon>0$, con la scelta $\delta=\varepsilon/3$: il limite vale $5$.

Nota la struttura: prima si fa un **conto preparatorio** ("lavoro sporco") per scoprire quale $\delta$ funziona, poi si scrive la **dimostrazione pulita** che parte direttamente dalla scelta di $\delta$ e verifica che funzioni. Questa distinzione (bozza vs. dimostrazione finale) è tipica di ogni argomento ε-δ.

### Dimostrare che un limite NON esiste

Usando la negazione del notebook 10.2: $f(x)$ **non** tende a $L$ per $x\to x_0$ significa $\exists\varepsilon>0\,\forall\delta>0\,\exists x\,(0<|x-x_0|<\delta\wedge|f(x)-L|\ge\varepsilon)$ — dobbiamo trovare **un** $\varepsilon$ che **nessun** $\delta$ riesce a soddisfare. Esempio: la funzione segno $\operatorname{sgn}(x)$ non ha limite per $x\to0$: scegliendo $\varepsilon=1$, per ogni $\delta>0$ esistono punti $x$ arbitrariamente vicini a $0$ (sia positivi che negativi) dove $\operatorname{sgn}(x)$ vale $+1$ o $-1$ — non più vicino di $1$ a **nessun** singolo candidato $L$.

````python
import sympy as sp

x = sp.symbols('x')
f = 3*x - 1
x0, L = 2, 5

# Verifica simbolica: per un dato epsilon, calcoliamo il delta = epsilon/3 e controlliamo
epsilon = sp.symbols('epsilon', positive=True)
delta = epsilon / 3

# Sostituiamo x = x0 + delta (il caso limite della disuguaglianza |x-x0| < delta)
espressione = sp.Abs(f.subs(x, x0 + delta) - L)
espressione_semplificata = sp.simplify(espressione)
print("Sostituendo x = x0 + delta, |f(x) - L| diventa:", espressione_semplificata)
print("Che è minore di epsilon per costruzione (delta = epsilon/3).")

# Verifica numerica su valori concreti di epsilon
import numpy as np
for eps in [0.1, 0.01, 0.001]:
    d = eps / 3
    x_test = x0 + d * 0.99  # un punto dentro l'intorno (0, delta)
    valore = 3*x_test - 1
    print(f"eps={eps}: delta={d:.5f}, x_test={x_test:.5f}, "
          f"|f(x_test)-L|={abs(valore-L):.6f} < eps? {abs(valore-L) < eps}")
````

## Esercizio svolto e commentato

**Problema:** dimostra con la definizione ε-δ che $\lim_{x\to3}x^2=9$.

**Ragionamento:**

**Passo 1 — lavoro preparatorio.** Vogliamo $|x^2-9|<\varepsilon$. Fattorizziamo (differenza di quadrati, notebook 1.2): $|x^2-9|=|x-3||x+3|$. Il fattore $|x-3|$ lo controlliamo con $\delta$, ma il fattore $|x+3|$ **dipende da $x$**, quindi va limitato a priori: se restringiamo la ricerca a $\delta\le1$, allora $|x-3|<1$ implica $2<x<4$, quindi $|x+3|<7$.

**Passo 2 — la stima combinata.** Con questa restrizione, $|x^2-9|=|x-3||x+3|<7|x-3|$. Vogliamo $7|x-3|<\varepsilon$, cioè $|x-3|<\varepsilon/7$. Quindi scegliamo $\delta=\min(1,\varepsilon/7)$ — il **minimo** tra i due vincoli, per soddisfare entrambe le richieste contemporaneamente (quella che garantisce $|x+3|<7$ e quella che garantisce la stima finale).

**Passo 3 — dimostrazione formale.** Dato $\varepsilon>0$, scegliamo $\delta=\min(1,\varepsilon/7)$. Se $0<|x-3|<\delta$: poiché $\delta\le1$, si ha $|x-3|<1$, quindi $2<x<4$ e $|x+3|<7$. Poiché $\delta\le\varepsilon/7$, si ha anche $|x-3|<\varepsilon/7$. Combinando: $|x^2-9|=|x-3||x+3|<\dfrac\varepsilon7\cdot7=\varepsilon$.

**Conclusione:** per ogni $\varepsilon>0$ esiste $\delta=\min(1,\varepsilon/7)>0$ che soddisfa la definizione: $\lim_{x\to3}x^2=9$. **Perché serve il "trucco del $\min$"**: a differenza dell'esempio lineare precedente, qui il rapporto incrementale non è costante — dipende da $x$ — e limitarlo richiede di restringere preventivamente il dominio di ricerca. Questa tecnica (limitare $\delta\le1$ per controllare un fattore variabile, poi stringere ulteriormente per la stima finale) è standard in ogni dimostrazione ε-δ con funzioni non lineari.

````python
import sympy as sp
import numpy as np

x = sp.symbols('x')
epsilon = sp.symbols('epsilon', positive=True)

delta = sp.Min(1, epsilon/7)

# Verifica numerica su una griglia di epsilon
for eps_val in [1.0, 0.1, 0.01]:
    d_val = min(1, eps_val/7)
    xs = np.linspace(3 - d_val + 1e-6, 3 + d_val - 1e-6, 2000)
    xs = xs[xs != 3]
    valori = xs**2 - 9
    massimo_scarto = np.max(np.abs(valori))
    print(f"eps={eps_val}: delta={d_val:.5f}, scarto massimo osservato={massimo_scarto:.6f}, "
          f"< eps? {massimo_scarto < eps_val}")
````

**1.** $|(2x+3)-5|=|2x-2|=2|x-1|$. Vogliamo $2|x-1|<\varepsilon$, cioè $|x-1|<\varepsilon/2$: scegliamo $\delta=\varepsilon/2$. Se $0<|x-1|<\delta$: $|(2x+3)-5|=2|x-1|<2\delta=\varepsilon$. Dimostrato.

**2.** $|x^2-0|=|x|^2$. Vogliamo $|x|^2<\varepsilon$, cioè $|x|<\sqrt\varepsilon$: scegliamo $\delta=\sqrt\varepsilon$. Se $0<|x-0|<\delta$: $|x^2-0|=|x|^2<\delta^2=\varepsilon$. Dimostrato (qui non serve il trucco del $\min$ perché non c'è un secondo fattore da limitare).

**3.** $|\sqrt x-2|=\left|\dfrac{(\sqrt x-2)(\sqrt x+2)}{\sqrt x+2}\right|=\dfrac{|x-4|}{\sqrt x+2}$. Restringendo $\delta\le1$ (quindi $3<x<5$, e $\sqrt x>\sqrt3>1$, dunque $\sqrt x+2>3$), si ottiene $\dfrac{|x-4|}{\sqrt x+2}<\dfrac{|x-4|}3$. Vogliamo $\dfrac{|x-4|}3<\varepsilon$, cioè $|x-4|<3\varepsilon$: scegliamo $\delta=\min(1,3\varepsilon)$. Se $0<|x-4|<\delta$: combinando le due stime, $|\sqrt x-2|<\dfrac{3\varepsilon}3=\varepsilon$. Dimostrato.

**4.** Fissiamo $\varepsilon=1$ (o qualsiasi $\varepsilon\le2$: mostreremo che nessun $L$ funziona con nessun $\delta$). Per ogni $\delta>0$, scegliamo $n$ abbastanza grande che $x_1=\dfrac1{\pi/2+2n\pi}$ e $x_2=\dfrac1{-\pi/2+2n\pi}$ soddisfino $0<|x_i|<\delta$ (possibile perché $x_1,x_2\to0$ per $n\to\infty$). Allora $\sin(1/x_1)=\sin(\pi/2+2n\pi)=1$ e $\sin(1/x_2)=\sin(-\pi/2+2n\pi)=-1$. Qualunque sia il candidato $L$, almeno uno tra $|1-L|$ e $|-1-L|$ è $\ge1$ (perché la distanza tra $1$ e $-1$ è $2$, quindi $L$ non può stare a distanza minore di $1$ da entrambi contemporaneamente). Quindi, con $\varepsilon=1$, nessun $\delta$ riesce a garantire $|\sin(1/x)-L|<\varepsilon$ per **tutti** gli $x$ nell'intorno: il limite non esiste.

Verifica computazionale (esercizio 4):

x1 = 1 / (np.pi/2 + 2*n*np.pi)
x2 = 1 / (-np.pi/2 + 2*n*np.pi)

## Riepilogo finale

- La definizione rigorosa di limite è $\forall\varepsilon>0\,\exists\delta>0:\,0<|x-x_0|<\delta\Rightarrow|f(x)-L|<\varepsilon$: il $\delta$ **può dipendere** da $\varepsilon$, e va trovato esplicitamente.
- Ogni dimostrazione ε-δ segue lo stesso schema: un **lavoro preparatorio** (manipolare $|f(x)-L|$ per capire quale $\delta$ serve) seguito dalla **dimostrazione pulita** (partire dalla scelta di $\delta$ e verificarla).
- Per funzioni non lineari serve spesso il **trucco del minimo**: restringere $\delta\le1$ per limitare un fattore variabile, poi stringere ulteriormente per la stima finale.
- Per dimostrare che un limite **non esiste**, si usa la negazione della definizione: trovare un $\varepsilon$ che nessun $\delta$ riesce a soddisfare.
