## Teoria, passo dopo passo

### Convergenza puntuale: la definizione, e il suo limite

Una successione di funzioni $f_n:E\to\mathbb R$ converge **puntualmente** a $f$ su $E$ se, per ogni $x\in E$ fissato, $f_n(x)\to f(x)$ (limite di successione numerica, notebook 10.10):

$$\forall x\in E\,\forall\varepsilon>0\,\exists N:\;n>N\Rightarrow|f_n(x)-f(x)|<\varepsilon$$

Qui $N$ **può dipendere sia da $\varepsilon$ sia da $x$** — esattamente come il $\delta$ nella continuità puntuale (notebook 10.13).

**Un esempio che rivela il problema**: $f_n(x)=x^n$ su $[0,1]$. Per $x\in[0,1)$, $x^n\to0$ (successione geometrica con ragione $|x|<1$, notebook 4.8); per $x=1$, $f_n(1)=1^n=1\to1$. Quindi $f_n\to f$ puntualmente, dove $f(x)=0$ per $x\in[0,1)$ e $f(1)=1$ — **una funzione discontinua**, benché **ogni** $f_n$ sia continua (un polinomio). La continuità **non è passata al limite**.

### Convergenza uniforme: la definizione

$$f_n\to f\text{ uniformemente su }E \quad\stackrel{\text{def}}{\Longleftrightarrow}\quad \forall\varepsilon>0\,\exists N:\;n>N\Rightarrow|f_n(x)-f(x)|<\varepsilon\;\;\forall x\in E$$

Lo stesso scambio di quantificatori visto nel notebook 10.13: ora $N$ dipende **solo** da $\varepsilon$, e deve funzionare per **tutti** i punti $x\in E$ simultaneamente. Geometricamente: da un certo $N$ in poi, **l'intero grafico** di $f_n$ resta dentro una "striscia" di ampiezza $2\varepsilon$ attorno al grafico di $f$ — non solo punto per punto, ma ovunque insieme.

**Perché $x^n$ non converge uniformemente su $[0,1]$**: fissato $\varepsilon=1/2$, per **ogni** $N$, scegliendo $x$ abbastanza vicino a $1$ (per continuità di $y\mapsto y^{N+1}$, esiste sempre $x<1$ con $x^{N+1}>1/2$), si ha $|f_{N+1}(x)-f(x)|=x^{N+1}>1/2=\varepsilon$ — nessun singolo $N$ riesce a "domare" simultaneamente tutti i punti vicino a $1$, dove la convergenza rallenta sempre di più.

### Il limite uniforme di funzioni continue è continuo

**Teorema.** Se $f_n\to f$ uniformemente su $E$, e ogni $f_n$ è continua in $x_0\in E$, allora $f$ è continua in $x_0$.

**Dimostrazione (il "trucco dei tre ε").** Fissato $\varepsilon>0$. Per convergenza uniforme, esiste $N$ tale che $|f_n(x)-f(x)|<\varepsilon/3$ per **ogni** $x\in E$ e ogni $n>N$; fissiamo un tale $n$. Per continuità di $f_n$ in $x_0$, esiste $\delta>0$ tale che $|x-x_0|<\delta\Rightarrow|f_n(x)-f_n(x_0)|<\varepsilon/3$. Per $|x-x_0|<\delta$:

$$|f(x)-f(x_0)|\le|f(x)-f_n(x)|+|f_n(x)-f_n(x_0)|+|f_n(x_0)-f(x_0)|<\frac\varepsilon3+\frac\varepsilon3+\frac\varepsilon3=\varepsilon$$

(disuguaglianza triangolare applicata due volte, inserendo e togliendo $f_n(x)$ e $f_n(x_0)$; il primo e il terzo termine usano la convergenza uniforme **valida per ogni $x$**, incluso $x_0$ stesso — è qui che l'uniformità, e non la sola convergenza puntuale, è essenziale). Dunque $f$ è continua in $x_0$. $\blacksquare$

**Perché fallisce con la sola convergenza puntuale**: nella dimostrazione, il primo termine $|f(x)-f_n(x)|<\varepsilon/3$ deve valere per **il punto $x$ che si avvicina a $x_0$**, non fissato in anticipo — con la sola convergenza puntuale, l'$N$ necessario per quel punto specifico potrebbe crescere senza controllo man mano che $x\to x_0$, vanificando la stima. Questo è esattamente ciò che accade con $x^n$: $x_0=1$, e i punti $x<1$ vicini a $1$ richiedono $N$ sempre più grandi.

### Integrazione e derivazione termine a termine

**Teorema (scambio con l'integrale, enunciato).** Se $f_n\to f$ uniformemente su $[a,b]$ e ogni $f_n$ è integrabile, allora $f$ è integrabile e $\int_a^bf_n\to\int_a^bf$ — cioè $\lim_n\int_a^bf_n=\int_a^b\lim_nf_n$.

L'idea della dimostrazione: $\left|\int_a^bf_n-\int_a^bf\right|\le\int_a^b|f_n-f|\le(b-a)\sup_{x\in[a,b]}|f_n(x)-f(x)|$, e quest'ultimo tende a $0$ **esattamente** per la definizione di convergenza uniforme (il $\sup$ su tutto l'intervallo che tende a $0$ è un modo equivalente di scrivere la convergenza uniforme). Con la sola convergenza puntuale questo scambio può fallire clamorosamente: è possibile costruire $f_n\to0$ puntualmente con $\int_0^1f_n=1$ per ogni $n$ (un "picco" che si stringe e si alza mantenendo area costante, mentre altrove tende a zero) — l'integrale del limite ($0$) non è il limite degli integrali ($1$). Un fenomeno analogo, ancora più delicato, vale per la derivata: la convergenza uniforme di $f_n$ **non basta** a garantire $f_n'\to f'$ — serve la convergenza uniforme delle **derivate stesse**.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 500)

fig, ax = plt.subplots(figsize=(7, 4.5))
for n in [1, 2, 5, 10, 30, 100]:
    ax.plot(x, x**n, label=f"n={n}")

ax.set_title("f_n(x) = x^n su [0,1]: convergenza puntuale a una funzione discontinua")
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

# Verifica numerica della non-uniformita': per ogni N, esiste x vicino a 1 con x^(N+1) > 1/2
def trova_x_problematico(N, target=0.5):
    x = 1.0
    passo = 0.5
    while x**(N+1) < target:
        x = 1 - (1 - x) / 2  # avvicina x a 1
    return x

for N in [5, 50, 500]:
    x_prob = trova_x_problematico(N)
    print(f"N={N}: x={x_prob:.8f} da x^(N+1)={x_prob**(N+1):.4f} > 0.5 "
          f"(nessun N doma TUTTI i punti vicino a 1)")
````

## Esercizio svolto e commentato

**Problema:** dimostra che $f_n(x)=\dfrac{\sin(nx)}n$ converge uniformemente a $f(x)=0$ su tutto $\mathbb R$, ma le derivate $f_n'(x)=\cos(nx)$ **non** convergono nemmeno puntualmente (in generale) — un esempio che chiarisce perché la derivazione termine a termine richiede ipotesi ulteriori.

**Ragionamento:**

**Passo 1 — stimiamo $|f_n(x)-f(x)|$ uniformemente.** $|f_n(x)-0|=\left|\dfrac{\sin(nx)}n\right|=\dfrac{|\sin(nx)|}n\le\dfrac1n$, usando $|\sin(nx)|\le1$ (limitatezza del seno, notebook 2.9) — **una stima che non dipende affatto da $x$**.

**Passo 2 — verifichiamo la convergenza uniforme.** Dato $\varepsilon>0$, per Archimede esiste $N$ con $\dfrac1N<\varepsilon$. Per $n>N$: $|f_n(x)-0|\le\dfrac1n<\dfrac1N<\varepsilon$, **per ogni $x\in\mathbb R$ simultaneamente** (lo stesso $N$ funziona ovunque, senza dover essere aggiustato punto per punto). Questo è precisamente la definizione di convergenza uniforme: $f_n\to0$ uniformemente su $\mathbb R$.

**Passo 3 — calcoliamo le derivate.** $f_n'(x)=\cos(nx)$ (regola di derivazione della funzione composta, notebook 5.9, applicata a $\sin(nx)/n$).

**Passo 4 — verifichiamo che $f_n'$ non converge, per esempio in $x=\pi$.** $f_n'(\pi)=\cos(n\pi)=(-1)^n$: la successione oscilla perennemente tra $+1$ e $-1$, non converge (stesso argomento delle sottosuccessioni divergenti visto nel notebook 10.7 con $(-1)^n$).

**Conclusione:** $f_n\to0$ uniformemente, ma $f_n'$ non converge nemmeno puntualmente in $x=\pi$ — quindi certamente non a $f'=0$. **La lezione**: la convergenza uniforme di $f_n$ da sola **non dice nulla** sul comportamento delle derivate $f_n'$; il "livellamento" delle oscillazioni verticali (piccola ampiezza $1/n$) non impedisce che le oscillazioni **orizzontali** (frequenza $n$, sempre più rapida) rendano le pendenze completamente instabili. Per derivare termine a termine serve, separatamente, che **le derivate stesse** convergano uniformemente.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 1000)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

for n in [2, 5, 20]:
    axes[0].plot(x, np.sin(n*x)/n, label=f"n={n}")
axes[0].set_title("f_n(x) = sin(nx)/n: ampiezza -> 0, converge uniformemente")
axes[0].legend()

for n in [2, 5, 20]:
    axes[1].plot(x, np.cos(n*x), label=f"n={n}", alpha=0.6)
axes[1].set_title("f_n'(x) = cos(nx): oscilla sempre più rapidamente, non converge")
axes[1].legend()

plt.tight_layout()
plt.show()

print("f_n'(pi) per n crescenti:", [round(np.cos(n*np.pi)) for n in range(1, 11)])
print("Oscilla perennemente tra 1 e -1: nessun limite.")
````

**1. Su $[0,1]$:** $|f_n(x)-0|=x/n\le1/n$ per ogni $x\in[0,1]$ (usando $x\le1$). **2.** $\dfrac{d}{dx}\left(\dfrac x{1+nx^2}\right)=\dfrac{(1+nx^2)-x\cdot2nx}{(1+nx^2)^2}=\dfrac{1-nx^2}{(1+nx^2)^2}$, nulla per $x=\pm1/\sqrt n$. In $x=1/\sqrt n$: $f_n(1/\sqrt n)=\dfrac{1/\sqrt n}{1+1}=\dfrac1{2\sqrt n}$. Questo è il massimo di $|f_n|$ su $\mathbb R$, e $\dfrac1{2\sqrt n}\to0$ per $n\to\infty$ (indipendentemente da $x$): dato $\varepsilon>0$, basta $n>1/(4\varepsilon^2)$ perché $\sup_x|f_n(x)|<\varepsilon$. Convergenza **uniforme** a $0$.

**3.** Ogni $f_n(x)=x^n$ è continua su $[0,1]$. Se la convergenza fosse uniforme, per il teorema il limite $f$ dovrebbe essere continuo su $[0,1]$. Ma $f$ (definita da $f(x)=0$ per $x<1$, $f(1)=1$) è discontinua in $x=1$ — contraddizione (contrapposizione della tesi del teorema). Dunque la convergenza non può essere uniforme.

**4.** Per la monotonia dell'integrale (se $g\le h$ ovunque, allora $\int g\le\int h$, conseguenza diretta della definizione con somme superiori/inferiori del notebook 10.15): $-\sup|f_n-f|\le f_n-f\le\sup|f_n-f|$ ovunque su $[a,b]$, quindi integrando: $-(b-a)\sup|f_n-f|\le\int_a^b(f_n-f)\le(b-a)\sup|f_n-f|$, cioè $\left|\int_a^bf_n-\int_a^bf\right|\le(b-a)\sup_{[a,b]}|f_n-f|$. Fissato $\varepsilon>0$, per convergenza uniforme esiste $N$ con $\sup_{[a,b]}|f_n-f|<\dfrac\varepsilon{b-a}$ per $n>N$ (assumendo $a\ne b$); allora $\left|\int_a^bf_n-\int_a^bf\right|<(b-a)\cdot\dfrac\varepsilon{b-a}=\varepsilon$. Dunque $\int_a^bf_n\to\int_a^bf$.

## Riepilogo finale

- La **convergenza puntuale** ($f_n(x)\to f(x)$ per ogni $x$ separatamente) non preserva la continuità: $x^n\to f$ discontinua su $[0,1]$.
- La **convergenza uniforme** richiede un unico $N$ valido per tutti i punti simultaneamente — lo stesso scambio di quantificatori della continuità uniforme (notebook 10.13).
- Il **limite uniforme di funzioni continue è continuo** (dimostrazione dei "tre ε"); la convergenza uniforme permette anche di scambiare limite e integrale.
- La convergenza uniforme di $f_n$ **non** garantisce nulla sulle derivate $f_n'$: $\sin(nx)/n\to0$ uniformemente, ma $\cos(nx)$ non converge nemmeno puntualmente.
