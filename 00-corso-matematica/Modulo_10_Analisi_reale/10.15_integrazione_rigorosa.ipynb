## Teoria, passo dopo passo

### Partizioni, somme superiori e inferiori

Una **partizione** di $[a,b]$ è un insieme finito di punti $P=\{a=x_0<x_1<\cdots<x_n=b\}$, che divide $[a,b]$ in sottointervalli $[x_{i-1},x_i]$. Sia $f:[a,b]\to\mathbb R$ **limitata**. Su ciascun sottointervallo, definiamo:

$$M_i=\sup_{x\in[x_{i-1},x_i]}f(x) \qquad\qquad m_i=\inf_{x\in[x_{i-1},x_i]}f(x)$$

(entrambi esistono per l'assioma di completezza, notebook 10.9, essendo $f$ limitata). La **somma superiore** e la **somma inferiore** relative a $P$ sono:

$$U(f,P)=\sum_{i=1}^nM_i(x_i-x_{i-1}) \qquad\qquad L(f,P)=\sum_{i=1}^nm_i(x_i-x_{i-1})$$

Geometricamente: $U(f,P)$ è l'area dei rettangoli che **contengono** il grafico su ogni sottointervallo (sovrastima), $L(f,P)$ quella dei rettangoli **contenuti sotto** il grafico (sottostima). Sempre $L(f,P)\le U(f,P)$ (perché $m_i\le M_i$ su ogni sottointervallo), e, raffinando la partizione (aggiungendo punti), le somme superiori **non crescono** e le somme inferiori **non decrescono** (restringere un sottointervallo può solo abbassare il suo $\sup$ locale o alzare il suo $\inf$ locale).

### Integrale superiore, integrale inferiore, integrabilità

$$\overline{\int_a^b}f=\inf_PU(f,P) \qquad\qquad \underline{\int_a^b}f=\sup_PL(f,P)$$

(l'estremo inferiore/superiore su **tutte** le partizioni possibili; esistono sempre per completezza, essendo $f$ limitata). Si dimostra sempre $\underline{\int_a^b}f\le\overline{\int_a^b}f$. $f$ è **integrabile secondo Riemann** se questi due valori **coincidono**, e in tal caso il valore comune è $\int_a^bf$.

### Il criterio di Riemann per l'integrabilità

**Teorema.** $f$ è integrabile su $[a,b]$ $\iff$ per ogni $\varepsilon>0$ esiste una partizione $P$ con $U(f,P)-L(f,P)<\varepsilon$.

**Dimostrazione ($\Leftarrow$).** Per ogni $\varepsilon>0$, esiste $P$ con $U(f,P)-L(f,P)<\varepsilon$. Poiché $L(f,P)\le\underline{\int_a^b}f\le\overline{\int_a^b}f\le U(f,P)$ (per definizione di $\sup$/$\inf$ su tutte le partizioni), si ha $0\le\overline{\int_a^b}f-\underline{\int_a^b}f\le U(f,P)-L(f,P)<\varepsilon$. Essendo $\varepsilon>0$ arbitrario, $\overline{\int_a^b}f-\underline{\int_a^b}f=0$ (se fosse $>0$, scegliendo $\varepsilon$ minore di quel valore si avrebbe una contraddizione): $f$ è integrabile. La direzione ($\Rightarrow$) è analoga, usando le definizioni di $\sup$/$\inf$ per avvicinarsi quanto si vuole ai due integrali con partizioni opportune.

### Ogni funzione continua è integrabile

**Teorema.** Se $f$ è continua su $[a,b]$, allora $f$ è integrabile su $[a,b]$.

**Dimostrazione.** Fissiamo $\varepsilon>0$. Per il teorema di Heine-Cantor (notebook 10.13), $f$ è **uniformemente continua** su $[a,b]$ (compatto): esiste $\delta>0$ tale che $|x-y|<\delta\Rightarrow|f(x)-f(y)|<\dfrac\varepsilon{b-a}$, per ogni $x,y\in[a,b]$.

Scegliamo una partizione $P$ con tutti i sottointervalli di lunghezza $<\delta$ (possibile: basta dividere $[a,b]$ in $n$ parti uguali con $n>\dfrac{b-a}\delta$, per Archimede). Su ogni $[x_{i-1},x_i]$, per il teorema di Weierstrass (notebook 10.12), $f$ raggiunge $M_i=f(u_i)$ e $m_i=f(v_i)$ per certi $u_i,v_i\in[x_{i-1},x_i]$. Poiché $|u_i-v_i|\le x_i-x_{i-1}<\delta$, l'uniforme continuità dà $M_i-m_i=f(u_i)-f(v_i)<\dfrac\varepsilon{b-a}$.

Quindi:

$$U(f,P)-L(f,P)=\sum_{i=1}^n(M_i-m_i)(x_i-x_{i-1})<\frac\varepsilon{b-a}\sum_{i=1}^n(x_i-x_{i-1})=\frac\varepsilon{b-a}\cdot(b-a)=\varepsilon$$

(la somma delle lunghezze dei sottointervalli è sempre $b-a$, telescopando). Per il criterio di Riemann appena dimostrato, $f$ è integrabile. $\blacksquare$

**Il ruolo, ora visibile, di ogni ipotesi**: la compattezza di $[a,b]$ garantisce Weierstrass (per definire $M_i,m_i$ come valori effettivamente raggiunti) *e* Heine-Cantor (per rendere $\delta$ indipendente dal sottointervallo, così che un **singolo** raffinamento della partizione controlli **tutti** gli scarti $M_i-m_i$ simultaneamente). Senza continuità uniforme, potremmo aver bisogno di partizioni sempre più fini vicino a certi punti e non ad altri, senza un controllo globale uniforme — esattamente il fenomeno patologico incontrato con $1/x$ nel notebook 10.13.

````python
import numpy as np

def f(x):
    return x**2

def somme_superiore_inferiore(f, a, b, n):
    punti = np.linspace(a, b, n + 1)
    U, L = 0.0, 0.0
    for i in range(n):
        xi_meno1, xi = punti[i], punti[i+1]
        campioni = np.linspace(xi_meno1, xi, 200)
        valori = f(campioni)
        U += valori.max() * (xi - xi_meno1)
        L += valori.min() * (xi - xi_meno1)
    return U, L

print("Somme superiori e inferiori per f(x)=x^2 su [0,1], al raffinare della partizione:")
for n in [4, 16, 64, 256]:
    U, L = somme_superiore_inferiore(f, 0, 1, n)
    print(f"n={n:4}: U={U:.6f}, L={L:.6f}, U-L={U-L:.6f}")

print("\nValore esatto dell'integrale (x^3/3 da 0 a 1):", 1/3)
print("U-L tende a 0: il criterio di Riemann per l'integrabilità è soddisfatto.")
````

## Esercizio svolto e commentato

**Problema:** usa il criterio di Riemann per dimostrare direttamente (senza invocare il teorema generale sulla continuità) che $f(x)=x$ è integrabile su $[0,1]$, e calcola $\int_0^1x\,dx$ dalle somme.

**Ragionamento:**

**Passo 1 — scegliamo la partizione uniforme.** $P_n=\{0,\tfrac1n,\tfrac2n,\ldots,1\}$, con $n$ sottointervalli di lunghezza $\tfrac1n$ ciascuno.

**Passo 2 — calcoliamo $M_i,m_i$.** $f(x)=x$ è crescente, quindi su $\left[\tfrac{i-1}n,\tfrac in\right]$ si ha $M_i=\tfrac in$ (valore all'estremo destro) e $m_i=\tfrac{i-1}n$ (valore all'estremo sinistro).

**Passo 3 — calcoliamo le somme.** $U(f,P_n)=\displaystyle\sum_{i=1}^n\frac in\cdot\frac1n=\frac1{n^2}\sum_{i=1}^ni=\frac1{n^2}\cdot\frac{n(n+1)}2=\frac{n+1}{2n}$ (usando la formula di somma dimostrata nell'esercizio 1 del notebook 10.4). Analogamente, $L(f,P_n)=\displaystyle\sum_{i=1}^n\frac{i-1}n\cdot\frac1n=\frac1{n^2}\sum_{i=0}^{n-1}i=\frac1{n^2}\cdot\frac{(n-1)n}2=\frac{n-1}{2n}$.

**Passo 4 — verifichiamo il criterio.** $U(f,P_n)-L(f,P_n)=\dfrac{n+1}{2n}-\dfrac{n-1}{2n}=\dfrac2{2n}=\dfrac1n$. Dato $\varepsilon>0$, per Archimede esiste $n$ con $\dfrac1n<\varepsilon$: il criterio di Riemann è soddisfatto, $f(x)=x$ è integrabile.

**Conclusione — il valore dell'integrale.** $U(f,P_n)=\dfrac{n+1}{2n}\to\dfrac12$ e $L(f,P_n)=\dfrac{n-1}{2n}\to\dfrac12$ per $n\to\infty$ (calcolo diretto di limiti di successioni razionali, notebook 4.2). Poiché $L(f,P_n)\le\int_0^1x\,dx\le U(f,P_n)$ per ogni $n$, e entrambi i lati tendono a $\dfrac12$, per il teorema del confronto (che dimostreremo in generale nel prossimo notebook come conseguenza dell'algebra dei limiti) $\int_0^1x\,dx=\dfrac12$ — coerente con la formula elementare $\dfrac{x^2}2\Big|_0^1$ vista nel notebook 5.16, ora ottenuta da zero, dalla definizione.

````python
def U_n(n):
    return (n + 1) / (2 * n)

def L_n(n):
    return (n - 1) / (2 * n)

print("n | U(f,P_n) | L(f,P_n) | U - L")
for n in [5, 50, 500, 5000]:
    print(f"{n:5} | {U_n(n):.6f} | {L_n(n):.6f} | {U_n(n) - L_n(n):.6f}")

print("\nEntrambe le successioni tendono a 1/2 =", 0.5)
````

**4.** Per la disuguaglianza triangolare "inversa" ($||u|-|v||\le|u-v|$, conseguenza della disuguaglianza triangolare ordinaria, notebook 0.8): per $x,y$ nello stesso sottointervallo, $|f(x)|-|f(y)|\le|f(x)-f(y)|\le M_i-m_i$ (quest'ultimo passo perché $f(x),f(y)\in[m_i,M_i]$). Passando al $\sup$ su $x$ e $y$ indipendentemente: $\sup|f|-\inf|f|\le M_i-m_i$ su ogni sottointervallo. Quindi, per ogni partizione $P$: $U(|f|,P)-L(|f|,P)\le U(f,P)-L(f,P)$. Poiché $f$ è integrabile, per ogni $\varepsilon>0$ esiste $P$ con $U(f,P)-L(f,P)<\varepsilon$ (criterio di Riemann); per la stessa $P$, $U(|f|,P)-L(|f|,P)\le U(f,P)-L(f,P)<\varepsilon$: il criterio di Riemann vale anche per $|f|$, dunque $|f|$ è integrabile.

## Riepilogo finale

- L'integrale di Riemann si costruisce con **somme superiori e inferiori**, $U(f,P)$ e $L(f,P)$, su partizioni finite: $f$ è integrabile se $\inf_PU(f,P)=\sup_PL(f,P)$.
- Il **criterio di Riemann** rende operativa questa definizione: basta trovare, per ogni $\varepsilon$, una partizione con $U(f,P)-L(f,P)<\varepsilon$.
- **Ogni funzione continua su $[a,b]$ è integrabile**: la dimostrazione usa Weierstrass (per definire $M_i,m_i$ come valori raggiunti) e Heine-Cantor (per un $\delta$ uniforme che controlla tutti gli scarti $M_i-m_i$ insieme).
- Non tutte le funzioni limitate sono integrabili: la funzione di Dirichlet è il controesempio canonico, con $U-L=1$ per ogni partizione, mai riducibile.
