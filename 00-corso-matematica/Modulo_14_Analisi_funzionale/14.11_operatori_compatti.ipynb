## Perché ci interessa

Nel notebook 12.6 abbiamo visto che, in dimensione finita, la palla unitaria chiusa è compatta (Heine-Borel). In dimensione infinita, questo **fallisce sempre**: la palla unitaria di uno spazio normato infinito-dimensionale non è mai compatta (fatto che dimostreremo). Gli **operatori compatti** sono quelli che, pur agendo su spazi di dimensione infinita, "comprimono" la palla unitaria in un insieme che **è** relativamente compatto nel codominio: si comportano, in un senso preciso, come se fossero "quasi di dimensione finita". Questa classe di operatori è centrale in analisi (equazioni integrali, teoria spettrale) proprio perché recupera, in un contesto ristretto, molte delle proprietà comode della dimensione finita.

## Teoria, passo dopo passo

### La palla unitaria non è mai compatta in dimensione infinita

**Teorema (di Riesz sulla compattezza).** Se $V$ è uno spazio normato di dimensione infinita, la palla unitaria chiusa $\overline{B}(0,1)=\{v:\|v\|\le1\}$ non è compatta.

**Dimostrazione (schema, usando il lemma di Riesz).** Il **lemma di Riesz** afferma: per ogni sottospazio chiuso proprio $M\subsetneq V$ e ogni $\theta\in(0,1)$, esiste $v\in V$ con $\|v\|=1$ e $\operatorname{dist}(v,M)\ge\theta$ (un vettore unitario "quasi ortogonale" a $M$, generalizzando l'ortogonalità esatta disponibile solo in spazi di Hilbert, notebook 14.3). Costruiamo induttivamente una successione $(v_n)$ con $\|v_n\|=1$ e $\|v_n-v_m\|\ge1/2$ per ogni $n\ne m$: partendo da un $v_1$ qualsiasi, dato $v_1,\ldots,v_{n-1}$, applichiamo il lemma di Riesz con $M=\operatorname{span}(v_1,\ldots,v_{n-1})$ (sottospazio di dimensione finita, quindi chiuso — un fatto vero in ogni spazio normato) e $\theta=1/2$, ottenendo $v_n$ con distanza $\ge1/2$ da ogni combinazione dei precedenti. Questa successione, essendo a distanza reciproca sempre $\ge1/2$, **non ammette alcuna sottosuccessione di Cauchy**, quindi nessuna sottosuccessione convergente: per la compattezza sequenziale (equivalente, in spazi metrici, alla compattezza per ricoprimenti, notebook 12.6), $\overline{B}(0,1)$ non è compatta. $\blacksquare$

**Contrasto con la dimensione finita.** In $\mathbb R^n$, Heine-Borel (notebook 10.11/12.6) garantisce che $\overline B(0,1)$ è compatta: questo teorema di Riesz mostra che la compattezza della palla unitaria è **equivalente** alla dimensione finita — una caratterizzazione sorprendentemente netta.

### Definizione di operatore compatto

Un operatore lineare $T:V\to W$ è **compatto** se $T(\overline B(0,1))$ è **relativamente compatto** in $W$ (cioè, la sua chiusura $\overline{T(\overline B(0,1))}$ è compatta, notebook 12.6) — equivalentemente, se ogni successione limitata $(v_n)$ in $V$ ha una sottosuccessione $(v_{n_k})$ tale che $(Tv_{n_k})$ converge in $W$.

**Ogni operatore compatto è limitato.** Se $T(\overline B(0,1))$ è relativamente compatto, è in particolare limitato (un insieme compatto in uno spazio metrico è sempre limitato, essendo unione finita di palle di raggio $1$ per il ricoprimento finito, notebook 12.6), quindi $\sup_{\|v\|\le1}\|Tv\|<\infty$: $T$ è limitato (notebook 14.5). Il viceversa è **falso**: l'identità $\operatorname{id}:V\to V$ è sempre limitata ($\|\operatorname{id}\|=1$, notebook 14.5), ma **non è compatta** se $\dim V=\infty$ (per il teorema di Riesz sopra, $\operatorname{id}(\overline B(0,1))=\overline B(0,1)$ non è relativamente compatta, essendo già essa stessa chiusa e non compatta).

### Esempio: operatori a rango finito sono compatti

Un operatore $T$ ha **rango finito** se $\dim(\operatorname{im}T)<\infty$. **Teorema.** Ogni operatore lineare limitato a rango finito è compatto.

**Dimostrazione.** $T(\overline B(0,1))$ è un sottoinsieme limitato (per la limitatezza di $T$) di $\operatorname{im}T$, uno spazio di dimensione **finita**. Per Heine-Borel (notebook 10.11/12.6, applicato allo spazio $\operatorname{im}T\cong\mathbb R^k$), un sottoinsieme limitato di $\operatorname{im}T$ ha chiusura compatta: $\overline{T(\overline B(0,1))}$ è compatta. $\blacksquare$

### Esempio: l'operatore integrale con nucleo continuo

Sia $K:C([0,1])\to C([0,1])$, $(Kf)(x)=\int_0^1k(x,y)f(y)\,dy$, con $k$ continua su $[0,1]\times[0,1]$ (notebook 12.9). **Teorema (enunciato).** $K$ è un operatore compatto. La dimostrazione (che omettiamo) usa il **teorema di Ascoli-Arzelà** (anch'esso enunciato senza dimostrazione in questo notebook introduttivo): la famiglia $\{Kf:\|f\|_\infty\le1\}$ è **equicontinua** (la continuità uniforme di $k$ su un compatto, notebook 10.13/12.6, garantisce un controllo uniforme dell'oscillazione di $Kf$) e uniformemente limitata, condizioni che Ascoli-Arzelà traduce in compattezza relativa in $C([0,1])$. Questo tipo di operatore, detto **operatore integrale**, è l'esempio guida storico che ha motivato l'intera teoria degli operatori compatti, nata dallo studio delle equazioni integrali di Fredholm.

```python
import numpy as np

# Verifica del lemma di Riesz "concettualmente": in dimensione infinita non si puo' avere
# una successione di vettori unitari tutti vicini tra loro come in dimensione finita

# Simuliamo con vettori canonici e_n in R^N per N grande, che approssimano il comportamento
# di uno spazio a dimensione "quasi infinita"
N = 50
vettori = np.eye(N)  # e_1, ..., e_N, tutti con norma 1

distanze = []
for i in range(N):
    for j in range(i+1, N):
        d = np.linalg.norm(vettori[i] - vettori[j])
        distanze.append(d)

print(f"Distanza tra ogni coppia di vettori canonici e_i, e_j (i!=j): {distanze[0]:.4f} (sempre sqrt(2))")
print(f"Minima distanza tra vettori distinti: {min(distanze):.4f}")
print("Nessuna sottosuccessione di (e_n) puo' convergere: la palla unitaria in dimensione alta")
print("contiene successioni 'sparse' che non si addensano mai -> non compatta se N->infinito")
```

## Esercizio svolto e commentato

**Problema:** dimostra che l'operatore diagonale $T:\ell^2\to\ell^2$, $T(x_1,x_2,x_3,\ldots)=(x_1,x_2/2,x_3/3,\ldots)$ (cioè $(Tx)_n=x_n/n$), è compatto, approssimandolo con operatori a rango finito.

**Ragionamento:**

**Passo 1 — costruiamo un'approssimazione a rango finito.** Definiamo $T_N(x)=(x_1,x_2/2,\ldots,x_N/N,0,0,\ldots)$ (tronca a zero dopo l'indice $N$): $T_N$ ha rango finito (immagine contenuta nel sottospazio delle prime $N$ coordinate, dimensione $\le N$), quindi è compatto per il teorema della teoria.

**Passo 2 — stimiamo $\|T-T_N\|$.** Per $x\in\ell^2$: $(T-T_N)(x)=(0,\ldots,0,x_{N+1}/(N+1),x_{N+2}/(N+2),\ldots)$. Dunque $\|(T-T_N)x\|_2^2=\sum_{n>N}x_n^2/n^2\le\dfrac1{(N+1)^2}\sum_{n>N}x_n^2\le\dfrac1{(N+1)^2}\|x\|_2^2$ (ogni termine della coda ha $1/n^2\le1/(N+1)^2$ per $n>N$). Dunque $\|(T-T_N)x\|_2\le\dfrac1{N+1}\|x\|_2$, cioè $\|T-T_N\|\le\dfrac1{N+1}\to0$ quando $N\to\infty$.

**Passo 3 — usiamo un fatto generale: il limite in norma di operatori compatti è compatto.** Questo fatto (che enunciamo senza dimostrare: l'insieme degli operatori compatti è chiuso in $\mathcal L(V,W)$, lo spazio normato degli operatori limitati introdotto nel notebook 14.5) dice che, essendo $T_N\to T$ in norma operatoriale (passo 2) e ogni $T_N$ compatto (passo 1), $T$ è anch'esso compatto.

**Conclusione:** $T$ è un operatore compatto su $\ell^2$, approssimato in norma da operatori a rango finito $T_N$ con errore che tende a $0$ come $1/N$. **Punto metodologico:** questa tecnica — approssimare con operatori a rango finito e stimare l'errore in norma — è il metodo standard per dimostrare la compattezza di un operatore diagonale (o, più in generale, di operatori "che decadono" lungo una base), ed è concettualmente la stessa idea usata per l'operatore integrale della teoria (dove l'approssimazione avviene tramite nuclei polinomiali o a variabili separate, invece che per troncamento diretto).

```python
import numpy as np

def T(x, n_max=100):
    n = np.arange(1, len(x)+1)
    return x / n

def T_N(x, N):
    y = T(x).copy()
    y[N:] = 0
    return y

np.random.seed(0)
x = np.random.randn(200)
x = x / np.linalg.norm(x)  # normalizziamo, ||x||_2 = 1

for N in [5, 20, 50, 100]:
    diff = T(x) - T_N(x, N)
    norma_diff = np.linalg.norm(diff)
    stima_teorica = 1 / (N + 1)
    print(f"N={N}: ||(T-T_N)(x)||_2 = {norma_diff:.5f}, stima teorica 1/(N+1) = {stima_teorica:.5f}")
```

## Riepilogo finale

- Il **teorema di Riesz** mostra che la palla unitaria è compatta se e solo se lo spazio ha dimensione finita: un fatto che segna un confine netto tra algebra lineare classica e analisi funzionale.
- Un **operatore compatto** manda la palla unitaria in un insieme relativamente compatto: è sempre limitato, ma non viceversa (l'identità in dimensione infinita è limitata ma mai compatta).
- Operatori a **rango finito** sono sempre compatti; l'operatore integrale con nucleo continuo (via Ascoli-Arzelà) è l'esempio storico guida.
- Un operatore compatto su uno spazio di dimensione infinita non può mai avere inversa limitata: la compattezza e l'invertibilità (in dimensione infinita) sono, in un certo senso, incompatibili.
