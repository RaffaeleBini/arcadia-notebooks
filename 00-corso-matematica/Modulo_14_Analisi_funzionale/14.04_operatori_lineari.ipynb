## Perché ci interessa

Nel notebook 6.13 un'applicazione lineare $T:\mathbb R^n\to\mathbb R^m$ era una funzione che rispetta somma e moltiplicazione per scalare, rappresentabile con una matrice (notebook 6.14). In dimensione infinita, la stessa idea di "rispettare la struttura lineare" si applica a funzioni tra spazi come $C([a,b])$ o $L^2$: la **derivata**, l'**integrale**, la **moltiplicazione per una funzione fissata** sono tutti esempi di operatori lineari, ma nessuno di essi è rappresentabile da una matrice finita. Questo notebook estende nucleo, immagine e le proprietà strutturali del Modulo 6 a questo contesto più ampio, preparando il terreno per la nozione di continuità (limitatezza) che tratteremo nel prossimo notebook.

## Teoria, passo dopo passo

### Definizione di operatore lineare

Siano $V,W$ spazi vettoriali (normati, in generale). Un **operatore lineare** $T:V\to W$ soddisfa $T(\alpha v_1+\beta v_2)=\alpha T(v_1)+\beta T(v_2)$ per ogni $v_1,v_2\in V$ e scalari $\alpha,\beta$ — identica alla definizione del notebook 6.13, ora su spazi di dimensione qualsiasi.

### Nucleo e immagine, generalizzati

Il **nucleo** $\ker T=\{v\in V:Tv=0\}$ e l'**immagine** $\operatorname{im}T=\{Tv:v\in V\}$ si definiscono esattamente come nel notebook 6.15, e restano sottospazi vettoriali di $V$ e $W$ rispettivamente (stessa dimostrazione: $0\in\ker T$ perché $T(0)=T(0\cdot v)=0\cdot T(v)=0$; chiusura per combinazioni lineari segue direttamente dalla linearità di $T$). **Ciò che cambia radicalmente in dimensione infinita**: il teorema del rango (notebook 6.15, $\dim\ker T+\dim\operatorname{im}T=\dim V$) non ha più senso quando $\dim V=\infty$ — le dimensioni infinite non si "sottraggono" in modo significativo, e questo teorema, pilastro dell'algebra lineare a dimensione finita, semplicemente non si applica qui.

### Esempi fondamentali di operatori su spazi di funzioni

- **Operatore di derivazione** $D:C^1([a,b])\to C([a,b])$, $Df=f'$. Lineare per le regole di derivazione (Modulo 5): $(αf+βg)'=αf'+βg'$. $\ker D=\{$funzioni costanti$\}$ (una funzione con derivata nulla ovunque è costante, teorema di Lagrange, notebook 10.14).
- **Operatore di integrazione** $I:C([a,b])\to C([a,b])$, $(If)(x)=\int_a^xf(t)\,dt$. Lineare per la linearità dell'integrale (Modulo 5/10.15). $\ker I=\{0\}$ (per il teorema fondamentale del calcolo, se $\int_a^xf=0$ per ogni $x$, derivando si ottiene $f=0$): $I$ è **iniettivo**.
- **Operatore di moltiplicazione** $M_g:L^2([a,b])\to L^2([a,b])$, $(M_gf)(x)=g(x)f(x)$, per $g\in C([a,b])$ fissata. Lineare per la distributiva (notebook 0.5) applicata puntualmente.
- **Operatore shift** su $\ell^2$ (successioni a quadrato sommabile, notebook 14.3): $S(x_1,x_2,x_3,\ldots)=(0,x_1,x_2,\ldots)$ (shift a destra). $\ker S=\{0\}$ ($S$ iniettivo), ma $\operatorname{im}S=\{y\in\ell^2:y_1=0\}\ne\ell^2$ ($S$ **non** suriettivo) — un fenomeno impossibile in dimensione finita, dove un operatore lineare iniettivo tra spazi della stessa dimensione è automaticamente suriettivo (conseguenza diretta del teorema del rango, notebook 6.15). In dimensione infinita, questa equivalenza **fallisce**.

### Operatori invertibili

Un operatore $T:V\to W$ è **invertibile** se esiste $T^{-1}:W\to V$ lineare con $T^{-1}T=\operatorname{id}_V$, $TT^{-1}=\operatorname{id}_W$ — equivalentemente, $T$ è biettivo (l'inversa di una funzione biettiva lineare è automaticamente lineare, verifica diretta identica al caso di dimensione finita). Come mostra l'esempio dello shift sopra, in dimensione infinita l'iniettività **non implica** l'invertibilità: serve verificare separatamente iniettività e suriettività (e, come vedremo nel notebook 14.5, spesso anche la continuità dell'inversa).

### L'algebra degli operatori: composizione e struttura vettoriale

L'insieme $\mathcal L(V,W)$ degli operatori lineari da $V$ a $W$ è esso stesso uno spazio vettoriale, con $(T_1+T_2)(v)=T_1v+T_2v$ e $(\alpha T)(v)=\alpha(Tv)$ (verifica diretta degli assiomi di spazio vettoriale, notebook 6.3). Quando $V=W$, $\mathcal L(V,V)$ ha anche un prodotto (la composizione $T_1\circ T_2$), rendendolo un'**algebra**: la composizione è associativa (proprietà generale della composizione di funzioni, notebook 3.7) e distributiva rispetto alla somma, ma **non commutativa** in generale — per esempio, $D\circ M_x\ne M_x\circ D$ sulle funzioni derivabili (dove $M_x$ moltiplica per $x$): $(D\circ M_x)(f)=(xf)'=f+xf'$, mentre $(M_x\circ D)(f)=xf'$, diversi in generale (a meno che $f=0$).

```python
import sympy as sp

x = sp.symbols('x')
f = sp.Function('f')(x)

# Verifica simbolica che D e M_x (moltiplicazione per x) non commutano
D_Mx_f = sp.diff(x * f, x)       # (D o M_x)(f) = d/dx(x*f)
Mx_D_f = x * sp.diff(f, x)       # (M_x o D)(f) = x*f'

print("D(M_x f) =", D_Mx_f)
print("M_x(D f) =", Mx_D_f)
print("Differenza:", sp.simplify(D_Mx_f - Mx_D_f))
print("(la differenza e' f stessa, non zero in generale: D e M_x NON commutano)")

# Verifica dello shift su una successione troncata
import numpy as np
x_vec = np.array([1, 2, 3, 4, 5])
shift = np.concatenate([[0], x_vec[:-1]])
print(f"\nx = {x_vec}")
print(f"S(x) = {shift} (shift a destra, la prima componente e' sempre 0)")
print("Nessuna successione con prima componente != 0 e' nell'immagine di S: S non e' suriettivo")
```

## Esercizio svolto e commentato

**Problema:** dimostra che l'operatore di integrazione $I:C([0,1])\to C([0,1])$, $(If)(x)=\int_0^xf(t)\,dt$, è lineare, calcola $\ker I$ e $\operatorname{im}I$, e verifica se $I$ è invertibile.

**Ragionamento:**

**Passo 1 — linearità.** $(I(\alpha f+\beta g))(x)=\int_0^x(\alpha f(t)+\beta g(t))\,dt=\alpha\int_0^xf(t)\,dt+\beta\int_0^xg(t)\,dt=\alpha(If)(x)+\beta(Ig)(x)$ (linearità dell'integrale, Modulo 5/10.15). Dunque $I(\alpha f+\beta g)=\alpha If+\beta Ig$: $I$ è lineare.

**Passo 2 — nucleo.** $If=0$ significa $\int_0^xf(t)\,dt=0$ per **ogni** $x\in[0,1]$. Derivando entrambi i membri (lecito, essendo $f$ continua, per il teorema fondamentale del calcolo, notebook 10.15): $f(x)=0$ per ogni $x$. Dunque $\ker I=\{0\}$: $I$ è **iniettivo**.

**Passo 3 — immagine.** Se $g=If$ per qualche $f\in C([0,1])$, allora $g(x)=\int_0^xf(t)\,dt$: per il teorema fondamentale del calcolo, $g$ è **derivabile** con $g'=f$ continua, e inoltre $g(0)=\int_0^0f=0$. Dunque $\operatorname{im}I\subseteq\{g\in C^1([0,1]):g(0)=0\}$. Viceversa, se $g\in C^1([0,1])$ con $g(0)=0$, ponendo $f=g'$ (continua, per ipotesi $g\in C^1$): $(If)(x)=\int_0^xg'(t)\,dt=g(x)-g(0)=g(x)$ (teorema fondamentale del calcolo), quindi $g=If\in\operatorname{im}I$. Dunque $\operatorname{im}I=\{g\in C^1([0,1]):g(0)=0\}$.

**Passo 4 — invertibilità.** $I$ non è suriettivo su $C([0,1])$: per esempio, la funzione costante $g(x)=1$ è continua ma non sta in $\operatorname{im}I$ (essendo $g(0)=1\ne0$, oppure semplicemente perché $g$ non è nemmeno della forma richiesta — inoltre nemmeno tutte le funzioni con $g(0)=0$ sono raggiunte se non sono $C^1$, come $g(x)=\sqrt x$, continua ma non derivabile in $0$). Dunque $I:C([0,1])\to C([0,1])$ **non è invertibile** (non essendo suriettivo), pur essendo iniettivo — lo stesso fenomeno dello shift della teoria.

**Conclusione:** $I$ è iniettivo ma non suriettivo su $C([0,1])$: un altro esempio concreto di come, in dimensione infinita, l'iniettività non implichi l'invertibilità, contrariamente a quanto accade sempre in dimensione finita (notebook 6.15).

```python
import numpy as np
from scipy import integrate

# Verifica numerica: I(f)(x) per f(t)=cos(t), confrontando con la primitiva nota
def f(t):
    return np.cos(t)

def I_f(x):
    return integrate.quad(f, 0, x)[0]

x_vals = np.linspace(0, 1, 10)
valori_I = [I_f(x) for x in x_vals]
valori_attesi = np.sin(x_vals)  # primitiva di cos che si annulla in 0

print("x\tI(f)(x)\tsin(x) (atteso)")
for x, v, a in zip(x_vals, valori_I, valori_attesi):
    print(f"{x:.3f}\t{v:.5f}\t{a:.5f}")

print(f"\nI(f)(0) = {I_f(0)} (sempre 0, coerente con l'immagine {{g: g(0)=0}})")
```

## Esercizi da risolvere

1. Verifica che l'operatore di derivazione $D:C^1([0,1])\to C([0,1])$ è lineare, e determina $\ker D$.
2. Dimostra che $D\circ I=\operatorname{id}_{C([0,1])}$ (derivando l'integrale si torna alla funzione di partenza), ma $I\circ D\ne\operatorname{id}_{C^1([0,1])}$ (suggerimento: calcola $(I\circ D)(f)$ per $f(x)=x+1$, usando il teorema fondamentale del calcolo).
3. Dimostra che se $g(x)\ne0$ per ogni $x\in[a,b]$, allora l'operatore di moltiplicazione $M_g$ (teoria) è iniettivo su $C([a,b])$. Mostra poi, con un esempio esplicito, che $M_g$ può restare iniettivo anche se $g$ si annulla in un **insieme finito** di punti (suggerimento: usa la continuità di $f$ per dedurre che $f$ deve annullarsi anche nei punti dove $g=0$, se $f$ è già forzata a $0$ altrove per densità).
4. *(più impegnativo)* Dimostra che l'operatore shift a **sinistra** $S^*:\ell^2\to\ell^2$, $S^*(x_1,x_2,x_3,\ldots)=(x_2,x_3,\ldots)$, è **suriettivo** ma **non iniettivo** (l'opposto dello shift a destra della teoria), calcolando esplicitamente $\ker S^*$.

**1.** $D(\alpha f+\beta g)=(\alpha f+\beta g)'=\alpha f'+\beta g'=\alpha Df+\beta Dg$ (regole di derivazione, Modulo 5). $\ker D=\{f\in C^1([0,1]):f'=0\text{ ovunque}\}$: per il teorema di Lagrange (notebook 10.14, corollario), una funzione con derivata identicamente nulla su un intervallo è costante. Dunque $\ker D=\{$funzioni costanti$\}$.

**2.** $(D\circ I)(f)(x)=\dfrac{d}{dx}\int_0^xf(t)\,dt=f(x)$ per il teorema fondamentale del calcolo (notebook 10.15): dunque $D\circ I=\operatorname{id}$. Per $f(x)=x+1$: $(I\circ D)(f)(x)=\int_0^xf'(t)\,dt=\int_0^x1\,dt=x$ (essendo $f'(t)=1$). Ma $f(x)=x+1\ne x=(I\circ D)(f)(x)$: dunque $I\circ D\ne\operatorname{id}$. **Spiegazione:** $I\circ D$ "dimentica" la costante additiva di $f$ (essendo $I$ costruito per dare sempre $0$ in $x=0$), mentre $D\circ I$ non ha questo problema perché la derivata elimina comunque le costanti.

**3.** Se $g(x)\ne0$ per ogni $x\in[a,b]$: $M_gf=0$ significa $g(x)f(x)=0$ per ogni $x$; essendo $g(x)\ne0$, deve essere $f(x)=0$ per ogni $x$, cioè $f=0$. Dunque $\ker M_g=\{0\}$: $M_g$ è iniettivo. **Esempio con zeri isolati:** sia $g(x)=x$ su $[-1,1]$ (si annulla solo in $x_0=0$). Se $M_gf=0$, cioè $x\cdot f(x)=0$ per ogni $x\in[-1,1]$: per $x\ne0$, essendo $x\ne0$, deve essere $f(x)=0$; per continuità di $f$ (notebook 12.4), $f(0)=\lim_{x\to0}f(x)=\lim_{x\to0}0=0$. Dunque $f=0$ ovunque, anche nel punto $x_0=0$ dove $g$ si annulla: $M_g$ resta iniettivo, pur avendo $g$ uno zero. Il meccanismo è la **continuità**: essendo $f$ forzata a $0$ in un insieme denso attorno a $0$ (tutti i punti $\ne0$), la continuità costringe $f$ a $0$ anche nel punto isolato mancante — un fenomeno che fallirebbe se si permettesse a $g$ di annullarsi su un **intero sottointervallo** (in tal caso $f$ potrebbe essere non nulla lì, mantenendo $gf=0$).

**4.** $S^*$ è suriettivo: dato $y=(y_1,y_2,\ldots)\in\ell^2$, basta prendere $x=(0,y_1,y_2,\ldots)$ (ancora in $\ell^2$, con $\|x\|_2=\|y\|_2$): $S^*(x)=(y_1,y_2,\ldots)=y$. $\ker S^*=\{x\in\ell^2:S^*(x)=0\}=\{x:(x_2,x_3,\ldots)=(0,0,\ldots)\}=\{(x_1,0,0,\ldots):x_1\in\mathbb R\}$: un sottospazio di dimensione $1$, non banale, quindi $S^*$ **non è iniettivo** (per esempio, $(1,0,0,\ldots)$ e $(2,0,0,\ldots)$ hanno entrambi $S^*=(0,0,\ldots)$).

## Riepilogo finale

- Un **operatore lineare** tra spazi vettoriali (anche di dimensione infinita) rispetta somma e moltiplicazione per scalare, esattamente come nel notebook 6.13; nucleo e immagine si definiscono allo stesso modo.
- Derivazione, integrazione, moltiplicazione per una funzione e shift su successioni sono esempi fondamentali di operatori su spazi infinito-dimensionali.
- Il **teorema del rango** (notebook 6.15) non ha senso in dimensione infinita: un operatore può essere iniettivo senza essere suriettivo (shift a destra, operatore di integrazione), o suriettivo senza essere iniettivo (shift a sinistra) — un'assenza di simmetria impossibile in dimensione finita.
- L'insieme degli operatori lineari $\mathcal L(V,W)$ è a sua volta uno spazio vettoriale, e $\mathcal L(V,V)$ è un'algebra non commutativa rispetto alla composizione.
