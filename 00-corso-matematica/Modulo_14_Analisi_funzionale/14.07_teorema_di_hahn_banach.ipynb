# 14.7 — Teorema di Hahn-Banach

**Prerequisiti:** 14.6 Funzionali, 14.1 Spazi normati.


## Perché ci interessa

Nel notebook 14.6 abbiamo visto che, su uno spazio di Hilbert, ogni funzionale continuo ha una forma esplicita (Riesz). Ma su uno spazio di Banach **generico** (senza prodotto scalare), non abbiamo strumenti così diretti per costruire funzionali. Il **teorema di Hahn-Banach** risolve un problema diverso ma altrettanto fondamentale: garantisce che un funzionale definito su un **sottospazio** possa sempre essere **esteso** a tutto lo spazio, mantenendo la stessa norma. Questo risultato, puramente esistenziale (non fornisce una formula, solo una garanzia), è il pilastro che assicura che ogni spazio normato non banale abbia "abbastanza" funzionali continui per essere utile — senza di esso, buona parte dell'analisi funzionale (compresa la dualità, la separazione di convessi, l'ottimizzazione) non potrebbe funzionare.

## Teoria, passo dopo passo

### Il problema dell'estensione

Sia $M\subseteq V$ un sottospazio di uno spazio normato $V$, e $\varphi:M\to\mathbb R$ un funzionale lineare continuo (rispetto alla norma ereditata da $V$). Vogliamo un funzionale $\Phi:V\to\mathbb R$ che **estenda** $\varphi$ (cioè $\Phi(m)=\varphi(m)$ per ogni $m\in M$) **senza aumentare la norma**: $\|\Phi\|_V=\|\varphi\|_M$ (dove i due sup sono presi rispettivamente su $V$ e su $M$).

**Perché non è ovvio.** Estendere una funzione lineare a uno spazio più grande è sempre possibile algebricamente (basta definirla arbitrariamente su un complemento algebrico di $M$, un fatto di puro algebra lineare); il problema è farlo **senza** far esplodere la norma. In dimensione infinita, dove basi algebriche (di Hamel) sono difficili da maneggiare esplicitamente, questo richiede un argomento non costruttivo.

### Enunciato del teorema di Hahn-Banach (versione per spazi normati)

**Teorema (Hahn-Banach).** Sia $V$ uno spazio normato, $M\subseteq V$ un sottospazio, $\varphi:M\to\mathbb R$ lineare con $|\varphi(m)|\le C\|m\|$ per ogni $m\in M$ (limitato con costante $C$). Allora esiste $\Phi:V\to\mathbb R$ lineare, con $\Phi|_M=\varphi$, e $|\Phi(v)|\le C\|v\|$ per ogni $v\in V$ (stessa costante $C$, quindi $\|\Phi\|=\|\varphi\|$).

**Idea della dimostrazione (schema, il caso generale usa il lemma di Zorn).** Si estende $\varphi$ **un elemento alla volta**: dato $v_0\notin M$, si costruisce un'estensione a $M\oplus\mathbb Rv_0$ (lo spazio generato da $M$ e $v_0$) definendo $\Phi(m+tv_0)=\varphi(m)+t\cdot\alpha$ per una costante $\alpha$ scelta con cura, in modo che la disuguaglianza $|\Phi(m+tv_0)|\le C\|m+tv_0\|$ resti soddisfatta per ogni $m\in M,t\in\mathbb R$ (un'analisi caso per caso, usando la disuguaglianza triangolare per trovare un intervallo non vuoto di valori ammissibili per $\alpha$ — il punto tecnico centrale, che omettiamo nei dettagli). Ripetendo questo processo, se $V/M$ ha dimensione finita, si estende $\varphi$ in un numero finito di passi (induzione, notebook 10.4). Per il caso generale (dimensione infinita, dove non basta un'induzione al finito), si usa il **lemma di Zorn** (un principio equivalente all'assioma della scelta) per garantire l'esistenza di un'estensione "massimale", che si dimostra essere definita su tutto $V$.

### Conseguenza 1: esistenza di funzionali non banali

**Corollario.** Per ogni $v_0\ne0$ in uno spazio normato $V$, esiste $\varphi\in V^*$ con $\|\varphi\|=1$ e $\varphi(v_0)=\|v_0\|$.

**Dimostrazione.** Sul sottospazio $M=\mathbb Rv_0$ (dimensione $1$), definiamo $\varphi_0(tv_0)=t\|v_0\|$: lineare, e $|\varphi_0(tv_0)|=|t|\|v_0\|=\|tv_0\|$, quindi $\|\varphi_0\|_M=1$ esattamente. Per Hahn-Banach, esiste un'estensione $\varphi$ a tutto $V$ con $\|\varphi\|=1$ e $\varphi(v_0)=\varphi_0(v_0)=\|v_0\|$. $\blacksquare$ Questo corollario garantisce che il duale $V^*$ **non è mai banale** (a meno che $V=\{0\}$): per ogni vettore non nullo, esiste un funzionale che lo "rileva" con la massima norma possibile.

### Conseguenza 2: il duale separa i punti

**Corollario.** Se $v_1\ne v_2$ in $V$, esiste $\varphi\in V^*$ con $\varphi(v_1)\ne\varphi(v_2)$.

**Dimostrazione.** Applicando il corollario precedente a $v_0=v_1-v_2\ne0$: esiste $\varphi$ con $\varphi(v_1-v_2)=\|v_1-v_2\|\ne0$, cioè $\varphi(v_1)\ne\varphi(v_2)$ (per linearità). $\blacksquare$ Questo fatto — "i funzionali continui distinguono i punti" — è l'analogo, per gli spazi normati, della proprietà di Hausdorff degli spazi topologici (notebook 12.6): garantisce che lo spazio duale sia "ricco abbastanza" da recuperare informazione su ogni singolo vettore.

```python
import numpy as np

# Illustrazione dell'estensione di Hahn-Banach in un caso concreto e finito-dimensionale
# M = sottospazio generato da (1,1,0) in R^3, phi definito su M
v0 = np.array([1, 1, 0])
norma_v0 = np.linalg.norm(v0)

# phi_0(t*v0) = t * ||v0||: definiamo phi_0 su M
def phi_0(t):
    return t * norma_v0

# Un'estensione naturale a R^3 che soddisfa Hahn-Banach: Phi(v) = v . (v0/||v0||)
u = v0 / norma_v0

def Phi(v):
    return np.dot(v, u)

# Verifica che Phi estende phi_0 su M
for t in [0.5, 1, 2, -1]:
    m = t * v0
    val_phi0 = phi_0(t)
    val_Phi = Phi(m)
    print(f"t={t}: phi_0(t*v0)={val_phi0:.4f}, Phi(t*v0)={val_Phi:.4f}, coincidono: {np.isclose(val_phi0, val_Phi)}")

# Verifica ||Phi|| = 1 (norma euclidea di u)
print(f"\n||u|| = {np.linalg.norm(u):.4f} (deve essere 1, cioe' ||Phi||=||phi_0||=1)")
```

## Esercizio svolto e commentato

**Problema:** su $V=\mathbb R^2$ con la norma euclidea, sia $M=\{(t,t):t\in\mathbb R\}$ (la retta $y=x$) e $\varphi_0(t,t)=2t$. Verifica che $\varphi_0$ è limitato su $M$ con una certa norma $\|\varphi_0\|_M$, e trova esplicitamente un'estensione $\Phi:\mathbb R^2\to\mathbb R$ che preserva questa norma.

**Ragionamento:**

**Passo 1 — calcoliamo $\|\varphi_0\|_M$.** Un generico elemento di $M$ è $(t,t)$, con norma euclidea $\|(t,t)\|=\sqrt{2t^2}=|t|\sqrt2$. $\varphi_0(t,t)=2t$, quindi $\dfrac{|\varphi_0(t,t)|}{\|(t,t)\|}=\dfrac{2|t|}{|t|\sqrt2}=\dfrac2{\sqrt2}=\sqrt2$ (costante, indipendente da $t$): $\|\varphi_0\|_M=\sqrt2$.

**Passo 2 — cerchiamo $\Phi(x,y)=ax+by$ (forma generale di un funzionale lineare su $\mathbb R^2$) che estenda $\varphi_0$.** Deve valere $\Phi(t,t)=at+bt=(a+b)t=2t$ per ogni $t$, cioè $a+b=2$.

**Passo 3 — determiniamo $a,b$ imponendo $\|\Phi\|=\sqrt2$.** Per un funzionale $\Phi(x,y)=ax+by=(x,y)\cdot(a,b)$, la norma è $\|\Phi\|=\|(a,b)\|=\sqrt{a^2+b^2}$ (caso particolare del notebook 14.6: il funzionale rappresentato dal vettore $(a,b)$ ha norma pari alla norma euclidea di quel vettore, per Cauchy-Schwarz con uguaglianza quando $(x,y)$ è proporzionale a $(a,b)$). Vogliamo $\sqrt{a^2+b^2}=\sqrt2$ insieme a $a+b=2$: sostituendo $b=2-a$ in $a^2+b^2=2$: $a^2+(2-a)^2=2\Rightarrow2a^2-4a+4=2\Rightarrow2a^2-4a+2=0\Rightarrow a^2-2a+1=0\Rightarrow(a-1)^2=0\Rightarrow a=1$, quindi $b=1$.

**Conclusione:** $\Phi(x,y)=x+y$ estende $\varphi_0$ (verifica: $\Phi(t,t)=t+t=2t=\varphi_0(t,t)$ ✓) con $\|\Phi\|=\sqrt{1^2+1^2}=\sqrt2=\|\varphi_0\|_M$: esattamente l'estensione garantita da Hahn-Banach. **Osservazione:** in questo esempio a dimensione finita, l'estensione risulta **unica** (l'equazione $(a-1)^2=0$ ha una sola soluzione) — un fatto **non garantito in generale** dal teorema di Hahn-Banach (che assicura solo l'**esistenza**, non l'unicità, dell'estensione): in dimensione infinita, spesso esistono infinite estensioni diverse che preservano la norma.

```python
import numpy as np

M_direction = np.array([1, 1]) / np.sqrt(2)  # direzione di M, normalizzata

def phi_0(t):
    return 2 * t

# Verifica della norma di phi_0 su M
for t in [1, 2, 5, -3]:
    punto = np.array([t, t])
    norma_punto = np.linalg.norm(punto)
    val = phi_0(t)
    print(f"t={t}: ||(" f"{t},{t})|| = {norma_punto:.4f}, phi_0 = {val}, rapporto = {abs(val)/norma_punto:.4f}")

# L'estensione trovata: Phi(x,y) = x+y
def Phi(x, y):
    return x + y

print(f"\n||Phi|| = ||(1,1)|| = {np.linalg.norm([1,1]):.4f} (atteso sqrt(2) = {np.sqrt(2):.4f})")
print(f"Phi(3,3) = {Phi(3,3)}, phi_0(3) = {phi_0(3)} (devono coincidere su M)")
```

## Riepilogo finale

- Il **teorema di Hahn-Banach** garantisce che un funzionale lineare limitato su un sottospazio si estende sempre a tutto lo spazio, **senza aumentare la norma** — un risultato di puro esistenza, dimostrato (nel caso generale) con il lemma di Zorn.
- **Conseguenza 1**: per ogni vettore non nullo esiste un funzionale di norma $1$ che lo "misura" esattamente alla sua norma — il duale non è mai banale.
- **Conseguenza 2**: i funzionali continui **separano i punti** dello spazio, analogamente alla proprietà di Hausdorff in topologia (notebook 12.6).
- L'estensione garantita da Hahn-Banach **non è unica** in generale (a differenza di casi particolari come spazi euclidei di dimensione finita): la non-unicità è più marcata quando la norma dello spazio è "meno rigida" (come $\|\cdot\|_\infty$ rispetto a $\|\cdot\|_2$).
