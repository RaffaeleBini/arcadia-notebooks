# 14.2 — Spazi di Banach

**Prerequisiti:** 14.1 Spazi normati, 12.8 Completezza (spazi metrici), 13.8 Spazi Lᵖ.


## Perché ci interessa

Nel notebook 12.8 abbiamo visto che uno spazio metrico può essere completo o no: $\mathbb R$ lo è, $\mathbb Q$ no. Per uno spazio normato (notebook 14.1), la completezza rispetto alla metrica indotta dalla norma è una proprietà **cruciale**: garantisce che ogni successione di Cauchy — costruita, per esempio, come somma parziale di una serie, o come approssimazione successiva in un metodo iterativo — converga effettivamente **dentro** lo spazio. Senza completezza, argomenti di esistenza (come i teoremi del punto fisso, o la costruzione stessa dell'integrale di Lebesgue nel notebook 13.4) non funzionerebbero. Gli spazi normati completi, chiamati **spazi di Banach**, sono lo scenario naturale per quasi tutta l'analisi funzionale.

## Teoria, passo dopo passo

### Definizione di spazio di Banach

Uno **spazio di Banach** è uno spazio normato $(V,\|\cdot\|)$ che è **completo** rispetto alla metrica indotta $d(v,w)=\|v-w\|$ (notebook 12.8): ogni successione di Cauchy in $V$ converge a un elemento di $V$.

### $\mathbb R^n$ è sempre uno spazio di Banach

Per il notebook 12.8, $(\mathbb R^n,\|\cdot\|_2)$ è completo. Essendo tutte le norme su $\mathbb R^n$ equivalenti (notebook 14.1, teorema enunciato), la completezza rispetto a **una** norma implica la completezza rispetto a **ogni** norma su $\mathbb R^n$ (una successione di Cauchy per una norma lo è per ogni norma equivalente, e converge allo stesso limite, essendo le topologie indotte identiche): $\mathbb R^n$ è uno spazio di Banach, qualunque sia la norma scelta.

### $L^p$ è uno spazio di Banach (Riesz-Fischer)

Il teorema di Riesz-Fischer, enunciato nel notebook 13.8, dice esattamente questo: $(L^p(X,\mu),\|\cdot\|_p)$ è completo per ogni $p\ge1$, quindi è uno spazio di Banach — il primo esempio **genuinamente infinito-dimensionale** di questa struttura.

### $C([a,b])$ con la norma del sup è uno spazio di Banach

**Teorema.** $(C([a,b]),\|\cdot\|_\infty)$ è completo.

**Dimostrazione.** Sia $(f_n)$ di Cauchy in $\|\cdot\|_\infty$: per ogni $\varepsilon>0$ esiste $N$ con $\|f_n-f_m\|_\infty<\varepsilon$ per $n,m\ge N$, cioè $\sup_x|f_n(x)-f_m(x)|<\varepsilon$. Per ogni $x$ fissato, $(f_n(x))_n$ è di Cauchy in $\mathbb R$ (essendo $|f_n(x)-f_m(x)|\le\|f_n-f_m\|_\infty<\varepsilon$), quindi converge (completezza di $\mathbb R$, notebook 10.10) a un valore che chiamiamo $f(x)$. Questo definisce una funzione limite $f$; resta da mostrare che $f_n\to f$ **uniformemente** (non solo puntualmente) e che $f$ è continua. Fissato $\varepsilon>0$ e $N$ come sopra, per $n\ge N$ e ogni $x$: facendo $m\to\infty$ nella disuguaglianza $|f_n(x)-f_m(x)|<\varepsilon$ (per $m,n\ge N$), si ottiene $|f_n(x)-f(x)|\le\varepsilon$ (i limiti preservano le disuguaglianze non strette, Modulo 10). Essendo questo vero per ogni $x$, $\|f_n-f\|_\infty\le\varepsilon$ per $n\ge N$: $f_n\to f$ uniformemente. Per il **teorema del limite uniforme di funzioni continue è continuo** (dimostrato rigorosamente nel notebook 10.16, con la tecnica dei tre epsilon), $f\in C([a,b])$. Dunque $(f_n)$ converge, in norma $\|\cdot\|_\infty$, a un elemento $f\in C([a,b])$: $C([a,b])$ è completo. $\blacksquare$

Questa dimostrazione **riusa integralmente** il lavoro del Modulo 10: la completezza di $C([a,b])$ non è un fatto nuovo da dimostrare da zero, ma la combinazione di due risultati già stabiliti (completezza di $\mathbb R$, e continuità del limite uniforme).

### Sottospazi chiusi di uno spazio di Banach sono di Banach

**Corollario.** Se $V$ è di Banach e $W\subseteq V$ è un sottospazio vettoriale **chiuso** (rispetto alla topologia della norma), allora $W$ è esso stesso uno spazio di Banach (con la norma ereditata).

**Dimostrazione.** Identica, punto per punto, all'esercizio 1 del notebook 12.8 (sottospazio chiuso di uno spazio metrico completo è completo): una successione di Cauchy in $W$ è di Cauchy in $V$, converge in $V$ per completezza di $V$, e il limite sta in $W$ perché $W$ è chiuso (notebook 12.3). $\blacksquare$

### Uno spazio normato incompleto: un avvertimento

Non ogni spazio normato "naturale" è di Banach. Esempio: lo spazio delle successioni **eventualmente nulle** (successioni reali con solo un numero finito di termini non nulli), con la norma $\|x\|_\infty=\sup_n|x_n|$, **non** è completo: la successione $x^{(k)}=(1,1/2,1/3,\ldots,1/k,0,0,\ldots)$ è di Cauchy (verifica diretta) ma "vorrebbe convergere" a $(1,1/2,1/3,\ldots)$, che ha **infiniti** termini non nulli, quindi non appartiene allo spazio di partenza — un fenomeno del tutto analogo a $\mathbb Q$ incompleto in $\mathbb R$ (notebook 12.8), ma qui nel contesto di spazi di funzioni/successioni.

```python
import numpy as np

# Verifica numerica: la successione x^(k) nello spazio delle successioni eventualmente nulle
# e' di Cauchy ma "esce" dallo spazio nel limite

def x_k(k, n_max=50):
    v = np.zeros(n_max)
    for i in range(k):
        v[i] = 1 / (i + 1)
    return v

# Verifica di Cauchy nella norma del sup
for k1, k2 in [(5, 10), (10, 20), (50, 100)]:
    diff = x_k(k1) - x_k(k2)
    norma_diff = np.max(np.abs(diff))
    print(f"||x^({k1}) - x^({k2})||_inf = {norma_diff:.5f} (atteso: 1/{min(k1,k2)+1} = {1/(min(k1,k2)+1):.5f})")

print("\nLa distanza tende a 0 (Cauchy), ma il 'limite' (1, 1/2, 1/3, ...) ha infiniti termini non nulli:")
print("non appartiene allo spazio delle successioni EVENTUALMENTE nulle -> spazio incompleto")
```

## Esercizio svolto e commentato

**Problema:** dimostra che lo spazio $\ell^\infty$ delle successioni **limitate** $x=(x_n)_n$, con norma $\|x\|_\infty=\sup_n|x_n|$, è uno spazio di Banach.

**Ragionamento:**

**Passo 1 — verifichiamo che $\|\cdot\|_\infty$ è una norma su $\ell^\infty$.** Identico, con $n$ al posto di $x$, alla verifica dell'esercizio 1 del notebook 14.1 per $C([a,b])$: positività, omogeneità e disuguaglianza triangolare seguono dalle stesse proprietà dell'estremo superiore.

**Passo 2 — sia $(x^{(k)})_k$ una successione di Cauchy in $\ell^\infty$.** Per ogni $\varepsilon>0$, esiste $K$ tale che $\|x^{(k)}-x^{(j)}\|_\infty<\varepsilon$ per $k,j\ge K$, cioè $\sup_n|x_n^{(k)}-x_n^{(j)}|<\varepsilon$.

**Passo 3 — per ogni indice $n$ fissato, la successione $(x_n^{(k)})_k$ è di Cauchy in $\mathbb R$.** Essendo $|x_n^{(k)}-x_n^{(j)}|\le\sup_m|x_m^{(k)}-x_m^{(j)}|<\varepsilon$ per $k,j\ge K$: per completezza di $\mathbb R$ (notebook 10.10), converge a un valore $y_n$. Definiamo $y=(y_n)_n$.

**Passo 4 — verifichiamo che $y\in\ell^\infty$ (è limitata).** Per $k$ fissato abbastanza grande (dal passo 2, con $j\to\infty$): $|x_n^{(k)}-y_n|\le\varepsilon$ per ogni $n$ (stesso argomento di passaggio al limite del notebook 14.2 per $C([a,b])$). Dunque $|y_n|\le|x_n^{(k)}|+\varepsilon\le\|x^{(k)}\|_\infty+\varepsilon$ per ogni $n$: $y$ è limitata (il membro destro non dipende da $n$), cioè $y\in\ell^\infty$.

**Passo 5 — verifichiamo che $x^{(k)}\to y$ in norma $\|\cdot\|_\infty$.** Dal passo 4, $\sup_n|x_n^{(k)}-y_n|\le\varepsilon$ per $k$ abbastanza grande, cioè $\|x^{(k)}-y\|_\infty\le\varepsilon$: essendo $\varepsilon$ arbitrario, $x^{(k)}\to y$ in $\ell^\infty$.

**Conclusione:** ogni successione di Cauchy in $\ell^\infty$ converge a un elemento di $\ell^\infty$: $\ell^\infty$ è uno spazio di Banach. **Confronto con l'esempio della teoria:** la differenza cruciale rispetto allo spazio delle successioni eventualmente nulle è che $\ell^\infty$ contiene **tutte** le successioni limitate, incluso il limite "naturale" di ogni successione di Cauchy in norma sup — non c'è nulla "fuori" verso cui la successione possa scappare.

```python
import numpy as np

# Verifica numerica su un esempio concreto di successione di Cauchy in ell^infinito
def x_k(k, n_max=20):
    # successione che si stabilizza componente per componente
    v = np.zeros(n_max)
    for i in range(n_max):
        v[i] = 1 - 1/(k + i + 1)  # ogni componente tende a 1 quando k cresce
    return v

for k1, k2 in [(5, 10), (20, 40), (100, 200)]:
    diff = np.max(np.abs(x_k(k1) - x_k(k2)))
    print(f"||x^({k1}) - x^({k2})||_inf = {diff:.5f}")

limite = np.ones(20)  # il limite atteso: tutte le componenti tendono a 1
print(f"\nLimite atteso y = (1,1,1,...), che e' limitata: appartiene a ell^infinito")
print(f"||x^(200) - y||_inf = {np.max(np.abs(x_k(200) - limite)):.6f}")
```

## Riepilogo finale

- Uno **spazio di Banach** è uno spazio normato completo rispetto alla metrica indotta dalla norma (notebook 12.8): $\mathbb R^n$, $L^p$ (Riesz-Fischer, notebook 13.8) e $C([a,b])$ con la norma del sup sono tutti esempi.
- La dimostrazione della completezza di $C([a,b])$ riusa direttamente il teorema del limite uniforme di funzioni continue (notebook 10.16): l'analisi funzionale costruisce sistematicamente sui risultati dei moduli precedenti.
- Sottospazi **chiusi** di uno spazio di Banach sono a loro volta di Banach; non tutti gli spazi normati "naturali" sono completi (esempio: le successioni eventualmente nulle).
- La convergenza assoluta di una serie ($\sum\|v_n\|<\infty$) implica la convergenza della serie stessa in uno spazio di Banach: una generalizzazione diretta del criterio di convergenza assoluta per le serie numeriche (Modulo 4/10).
