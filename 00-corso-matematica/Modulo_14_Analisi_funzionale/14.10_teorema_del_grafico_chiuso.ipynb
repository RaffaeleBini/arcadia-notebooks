# 14.10 — Teorema del grafico chiuso

**Prerequisiti:** 14.9 Teorema dell'applicazione aperta, 12.9 Spazi prodotto, 12.3 Insiemi aperti e chiusi.


## Perché ci interessa

Verificare direttamente che un operatore lineare $T$ è limitato (notebook 14.5) richiede di stimare $\|Tv\|$ in funzione di $\|v\|$ per **ogni** $v$ contemporaneamente — spesso un compito delicato. Il **teorema del grafico chiuso**, il terzo e ultimo grande teorema basato sul teorema di Baire in questo modulo, offre un criterio spesso più maneggevole: basta verificare che il **grafico** di $T$ sia chiuso (una condizione che si controlla una successione alla volta, separando l'esistenza dei due limiti coinvolti). Sorprendentemente, per operatori lineari tra spazi di Banach, questa condizione apparentemente più debole è **equivalente** alla limitatezza.

## Teoria, passo dopo passo

### Il grafico di un operatore

Il **grafico** di $T:V\to W$ è $\Gamma_T=\{(v,Tv):v\in V\}\subseteq V\times W$ (con la topologia prodotto, notebook 12.9). $\Gamma_T$ è **chiuso** se: ogniqualvolta $(v_n,Tv_n)\to(v,w)$ in $V\times W$ (cioè $v_n\to v$ **e** $Tv_n\to w$, per il notebook 12.9), segue che $w=Tv$.

**Osservazione preliminare.** Se $T$ è continuo, $\Gamma_T$ è sempre chiuso: da $v_n\to v$ segue $Tv_n\to Tv$ (continuità), e se anche $Tv_n\to w$, per l'unicità del limite (Modulo 10) $w=Tv$. La direzione **interessante e non ovvia** del teorema è il viceversa.

### Enunciato del teorema del grafico chiuso

**Teorema (del grafico chiuso).** Siano $V,W$ spazi di Banach, e $T:V\to W$ un operatore **lineare** (non necessariamente assunto continuo) con grafico $\Gamma_T$ chiuso in $V\times W$. Allora $T$ è limitato (continuo).

**Dimostrazione (schema, usando il teorema dell'inversa limitata).** $\Gamma_T$, essendo un sottospazio vettoriale chiuso di $V\times W$ (verifica diretta: combinazioni lineari di elementi del grafico stanno nel grafico, per linearità di $T$), è esso stesso uno spazio di Banach (sottospazio chiuso di uno spazio di Banach, notebook 14.2, dove $V\times W$ è di Banach con la norma $\|(v,w)\|=\|v\|_V+\|w\|_W$, verifica diretta degli assiomi). Consideriamo le proiezioni $\pi_V:\Gamma_T\to V$, $\pi_V(v,Tv)=v$, e $\pi_W:\Gamma_T\to W$, $\pi_W(v,Tv)=Tv$: entrambe lineari e limitate (essendo $\|\pi_V(v,Tv)\|=\|v\|\le\|v\|+\|Tv\|=\|(v,Tv)\|$, e analogamente per $\pi_W$). $\pi_V$ è **biettiva** da $\Gamma_T$ a $V$ (per ogni $v\in V$, esiste l'unico punto $(v,Tv)\in\Gamma_T$ che vi si proietta): per il teorema dell'inversa limitata (notebook 14.9, applicabile perché sia $\Gamma_T$ sia $V$ sono di Banach), $\pi_V^{-1}:V\to\Gamma_T$, $\pi_V^{-1}(v)=(v,Tv)$, è limitata. Dunque $T=\pi_W\circ\pi_V^{-1}$ (composizione di $v\mapsto(v,Tv)\mapsto Tv$) è composizione di operatori limitati, quindi limitata (notebook 14.5, esercizio 3). $\blacksquare$

**Il punto tecnico cruciale.** Questa dimostrazione **riusa** il teorema dell'inversa limitata (notebook 14.9), che a sua volta si basa sul teorema dell'applicazione aperta e, in definitiva, su Baire (notebook 14.8): i tre grandi teoremi di questa parte del modulo sono intimamente collegati, ciascuno costruito sul precedente.

### Perché il criterio è più facile da verificare in pratica

Verificare direttamente la limitatezza richiede di controllare $\|Tv\|\le C\|v\|$ **simultaneamente** per ogni $v$. Verificare che il grafico sia chiuso richiede invece **due passi separati**: (1) assumere che $v_n\to v$ **e** $Tv_n\to w$ (entrambe le convergenze **date**, non da dimostrare), e (2) dedurre $w=Tv$. Questo è spesso più facile perché si ha già a disposizione **entrambi** i limiti come ipotesi, invece di dover produrre una stima uniforme da zero — una situazione comune quando $T$ è definito tramite equazioni differenziali o integrali, dove la convergenza di $Tv_n$ è spesso più facile da stabilire *a priori* insieme a quella di $v_n$, piuttosto che dedurne una stima quantitativa diretta.

```python
import numpy as np

# Illustrazione concettuale: un operatore su R^2 (dove ogni operatore lineare e' automaticamente continuo,
# quindi il grafico e' sempre chiuso) per visualizzare la nozione di grafico chiuso
A = np.array([[1, 2], [3, 1]])

def T(v):
    return A @ v

# Generiamo una successione v_n -> v e verifichiamo che (v_n, T(v_n)) -> (v, T(v))
v = np.array([1.0, 2.0])
v_n_list = [v + np.array([1/n, -1/n]) for n in range(1, 20)]

print("v_n -> v = (1,2), verifichiamo (v_n, T(v_n)) -> (v, T(v)):")
for n, v_n in zip([1, 5, 10, 19], [v_n_list[0], v_n_list[4], v_n_list[9], v_n_list[18]]):
    Tv_n = T(v_n)
    print(f"n={n}: v_n={v_n.round(4)}, T(v_n)={Tv_n.round(4)}")

Tv = T(v)
print(f"\nLimite atteso: v={v}, T(v)={Tv}")
print("Il grafico e' chiuso: la convergenza di v_n insieme a quella di T(v_n) forza il limite corretto")
```

## Esercizio svolto e commentato

**Problema:** usa il teorema del grafico chiuso per dimostrare che, se $T:H\to H$ è un operatore lineare su uno spazio di Hilbert (notebook 14.3) che soddisfa $\langle Tv,w\rangle=\langle v,Tw\rangle$ per ogni $v,w$ (cioè $T$ è **simmetrico**), allora $T$ è automaticamente limitato (un fatto notevole: la simmetria, una condizione puramente algebrica, implica la continuità).

**Ragionamento:**

**Passo 1 — impostiamo la verifica del grafico chiuso.** Supponiamo $v_n\to v$ e $Tv_n\to w$ in $H$; dobbiamo dimostrare $w=Tv$.

**Passo 2 — usiamo la simmetria per riscrivere il prodotto scalare con un $u$ di test arbitrario.** Per ogni $u\in H$: $\langle Tv_n,u\rangle=\langle v_n,Tu\rangle$ (simmetria di $T$, applicata con ruoli scambiati).

**Passo 3 — passiamo al limite in $n$ su entrambi i lati.** Il prodotto scalare è continuo rispetto a ciascun argomento (conseguenza di Cauchy-Schwarz, notebook 14.3: $|\langle a_n,b\rangle-\langle a,b\rangle|=|\langle a_n-a,b\rangle|\le\|a_n-a\|\|b\|\to0$ se $a_n\to a$). Dunque, usando $Tv_n\to w$ a sinistra e $v_n\to v$ a destra: $\langle w,u\rangle=\lim_n\langle Tv_n,u\rangle=\lim_n\langle v_n,Tu\rangle=\langle v,Tu\rangle$.

**Passo 4 — usiamo di nuovo la simmetria e concludiamo.** $\langle v,Tu\rangle=\langle Tv,u\rangle$ (simmetria, applicata direttamente). Dunque $\langle w,u\rangle=\langle Tv,u\rangle$ per **ogni** $u\in H$, cioè $\langle w-Tv,u\rangle=0$ per ogni $u$; scegliendo $u=w-Tv$: $\|w-Tv\|^2=0$, quindi $w=Tv$ (non-degenerazione del prodotto scalare, notebook 14.3).

**Conclusione:** il grafico di $T$ è chiuso (essendo verificata la condizione $w=Tv$ ogniqualvolta $v_n\to v$ e $Tv_n\to w$). Per il teorema del grafico chiuso (essendo $H$ di Hilbert, quindi di Banach, notebook 14.3), $T$ è automaticamente limitato. **Portata del risultato:** questo teorema (noto come teorema di Hellinger-Toeplitz) è un esempio paradigmatico dell'uso del grafico chiuso: la simmetria, condizione puramente algebrica su prodotti scalari, non menziona affatto la continuità, eppure la implica **automaticamente** grazie a questo criterio — un fatto sorprendente, spesso citato come giustificazione dell'importanza pratica del teorema.

```python
import numpy as np

# Verifica numerica su una matrice simmetrica (che rappresenta un operatore simmetrico su R^n, gia' finito-dim)
A = np.array([[2, 1, 0], [1, 3, 1], [0, 1, 1]])  # matrice simmetrica
print("A e' simmetrica?", np.allclose(A, A.T))

v = np.array([1.0, 2.0, -1.0])
w = np.array([0.5, -1.0, 2.0])

lhs = np.dot(A @ v, w)  # <Tv, w>
rhs = np.dot(v, A @ w)  # <v, Tw>
print(f"<Tv,w> = {lhs:.4f}, <v,Tw> = {rhs:.4f}, coincidono: {np.isclose(lhs, rhs)}")
print("\nIn dimensione finita ogni operatore e' automaticamente limitato (notebook 6.13);")
print("il teorema di Hellinger-Toeplitz e' interessante proprio perche' vale anche in dimensione infinita.")
```

## Riepilogo finale

- Il **grafico** di un operatore $T$ è $\Gamma_T=\{(v,Tv)\}\subseteq V\times W$; è sempre chiuso se $T$ è continuo, ma il viceversa è il contenuto non banale del teorema.
- Il **teorema del grafico chiuso** garantisce che un operatore lineare tra spazi di Banach con grafico chiuso è automaticamente limitato — dimostrato applicando il teorema dell'inversa limitata (notebook 14.9) alla proiezione $\pi_V:\Gamma_T\to V$.
- Il criterio del grafico chiuso è spesso più facile da verificare, perché richiede solo di dedurre $w=Tv$ da **due limiti già dati**, invece di produrre una stima uniforme diretta.
- Il **teorema di Hellinger-Toeplitz** (esercizio svolto) mostra un'applicazione sorprendente: un operatore simmetrico su uno spazio di Hilbert è automaticamente limitato, un fatto puramente algebrico con conseguenza topologica non ovvia. Con questo si chiude il tris di teoremi fondati sul teorema di Baire (Banach-Steinhaus, applicazione aperta, grafico chiuso).
