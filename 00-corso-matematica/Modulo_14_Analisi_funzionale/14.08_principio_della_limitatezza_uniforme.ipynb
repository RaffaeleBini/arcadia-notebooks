# 14.8 — Principio della limitatezza uniforme

**Prerequisiti:** 14.5 Operatori continui, 14.2 Spazi di Banach.


## Perché ci interessa

Nel notebook 14.5 abbiamo studiato la limitatezza di un **singolo** operatore. Qui affrontiamo una domanda diversa: data una **famiglia** (anche infinita) di operatori limitati, se ciascuno di essi manda ogni singolo vettore in un insieme limitato (limitatezza **puntuale**), possiamo concludere che le loro norme sono **uniformemente** limitate (un solo numero che limita tutte le norme insieme)? La risposta, sorprendente, è **sì** — ma solo se il dominio è uno spazio di Banach. Questo è il **principio della limitatezza uniforme** (o teorema di Banach-Steinhaus), il primo dei tre grandi teoremi dell'analisi funzionale che sfruttano la completezza in modo essenziale e non ovvio, tramite il teorema di categoria di Baire.

## Teoria, passo dopo passo

### Il teorema di categoria di Baire (enunciato, strumento tecnico)

**Teorema (Baire).** Se $(X,d)$ è uno spazio metrico completo (notebook 12.8) e $X=\bigcup_{n=1}^\infty F_n$ con ogni $F_n$ chiuso, allora almeno uno degli $F_n$ ha **interno non vuoto** (notebook 12.3).

Questo teorema — che omettiamo di dimostrare in dettaglio, essendo tecnico ma concettualmente affine agli argomenti di intervalli incapsulati del notebook 10.9 — dice che uno spazio completo **non può essere "esaurito"** da un'unione numerabile di chiusi "sottili" (senza punti interni, detti **rari** o **nowhere dense**). È lo strumento tecnico dietro tutti e tre i grandi teoremi di questo tratto del modulo.

### Enunciato del principio della limitatezza uniforme

**Teorema (Banach-Steinhaus).** Sia $V$ uno spazio di Banach, $W$ uno spazio normato, e $\{T_i\}_{i\in I}$ una famiglia di operatori lineari limitati $T_i:V\to W$. Se per ogni $v\in V$ l'insieme $\{T_iv:i\in I\}$ è limitato in $W$ (**limitatezza puntuale**: $\sup_i\|T_iv\|<\infty$ per ogni $v$ fissato), allora $\sup_i\|T_i\|<\infty$ (**limitatezza uniforme** delle norme).

**Dimostrazione (schema, usando Baire).** Per ogni $n$, poniamo $F_n=\{v\in V:\|T_iv\|\le n\text{ per ogni }i\in I\}$. Ogni $F_n$ è chiuso (intersezione di insiemi chiusi $\{v:\|T_iv\|\le n\}$, ciascuno controimmagine di un chiuso tramite $T_i$ continuo, notebook 12.4). Per l'ipotesi di limitatezza puntuale, ogni $v\in V$ sta in qualche $F_n$ (essendo $\sup_i\|T_iv\|<\infty$, esiste $n$ che lo supera): $V=\bigcup_nF_n$. Per Baire (essendo $V$ completo), qualche $F_{n_0}$ ha interno non vuoto: esiste una palla $B(v_0,r)\subseteq F_{n_0}$. Dunque, per ogni $v$ con $\|v\|\le r$ (cioè $v_0+v\in B(v_0,r)$) e ogni $i$: $\|T_i(v_0+v)\|\le n_0$, e anche $\|T_iv_0\|\le n_0$ (essendo $v_0\in F_{n_0}$); per la disuguaglianza triangolare, $\|T_iv\|=\|T_i(v_0+v)-T_iv_0\|\le\|T_i(v_0+v)\|+\|T_iv_0\|\le2n_0$. Dunque, per $\|v\|\le r$: $\|T_iv\|\le2n_0$ per ogni $i$, quindi $\|T_i\|=\sup_{\|v\|=r}\|T_iv\|/r\le2n_0/r$ (riscalando): una costante **indipendente da $i$**. $\blacksquare$

**Perché serve la completezza.** L'ipotesi $V$ di Banach è essenziale: è l'ingrediente che rende applicabile il teorema di Baire. Su uno spazio normato incompleto (come le successioni eventualmente nulle, notebook 14.2), il teorema può fallire — esistono controesempi espliciti (che omettiamo) dove famiglie puntualmente limitate non sono uniformemente limitate.

### Applicazione: divergenza delle serie di Fourier in un punto

**Applicazione classica (enunciata).** Il principio della limitatezza uniforme è usato (in un argomento più elaborato, che qui solo accenniamo) per dimostrare l'**esistenza** di funzioni continue la cui serie di Fourier diverge in almeno un punto: si definiscono operatori $T_n(f)=S_n(f)(0)$ (la somma parziale $n$-esima della serie di Fourier di $f$, valutata in $0$) su $C(\mathbb T)$ (funzioni continue periodiche); si dimostra che $\sup_n\|T_n\|=\infty$ (le norme **non** sono uniformemente limitate, calcolabile esplicitamente); per il **contronominale** del teorema di Banach-Steinhaus, deve esistere almeno una $f$ per cui $\{T_n(f)\}_n$ **non** è limitata (altrimenti, se lo fosse per ogni $f$ — limitatezza puntuale — il teorema garantirebbe la limitatezza uniforme delle norme, contraddicendo il calcolo). Questo è un tipico "argomento di esistenza senza costruzione esplicita": il teorema garantisce che una tale $f$ patologica esiste, senza indicarne una.

```python
import numpy as np
import matplotlib.pyplot as plt

# Illustriamo il principio con un esempio concreto: operatori T_n(x) = n * x_n su una successione,
# limitati puntualmente ma NON uniformemente, su uno spazio NON completo (successioni eventualmente nulle)
# per mostrare cosa succederebbe SENZA la completezza

def T_n_applicato(x, n):
    # x e' una successione (array), T_n(x) = n-esima componente moltiplicata per n
    if n-1 < len(x):
        return n * x[n-1]
    return 0

# Su successioni eventualmente nulle, x_k ha solo le prime k componenti diverse da 0
def x_k(k, n_max=50):
    v = np.zeros(n_max)
    if k-1 < n_max:
        v[k-1] = 1/k  # l'unica componente non nulla e' la k-esima, con valore 1/k
    return v

# Per ogni x fissato (a supporto finito), T_n(x) e' zero per n oltre il supporto: limitato puntualmente
x_test = x_k(5)
valori_T_n = [T_n_applicato(x_test, n) for n in range(1, 10)]
print(f"x_test (supporto in posizione 5, valore 1/5): T_n(x_test) per n=1..9: {valori_T_n}")
print("Limitato puntualmente su x_test (diventa 0 dopo n=5)")

# Ma la norma di T_n come operatore e' n stesso (su un vettore concentrato in posizione n)
print("\n||T_n|| = n (applicato al vettore e_n, con x_n=1): NON limitato uniformemente")
print("Questo NON contraddice Banach-Steinhaus, perche' lo spazio delle successioni")
print("eventualmente nulle NON e' completo (notebook 14.2): l'ipotesi del teorema fallisce")
```

## Esercizio svolto e commentato

**Problema:** su $\ell^2$ (spazio di Banach, in realtà di Hilbert, notebook 14.3), sia $T_n:\ell^2\to\mathbb R$, $T_n(x)=\sqrt n\,x_n$ (l'operatore che isola la $n$-esima componente, amplificata da $\sqrt n$). Dimostra che $\sup_n\|T_n\|=\infty$, poi usa il contronominale di Banach-Steinhaus per dedurre l'esistenza di un $x\in\ell^2$ con $\sup_n|T_n(x)|=\infty$, e costruisci esplicitamente un tale $x$.

**Ragionamento:**

**Passo 1 — calcoliamo $\|T_n\|$ per $n$ fissato.** Per Cauchy-Schwarz applicata a $T_n(x)=\sqrt n\,x_n$: essendo $x_n^2\le\sum_mx_m^2=\|x\|_2^2$ (un singolo termine non negativo è al più la somma di tutti), $|x_n|\le\|x\|_2$, quindi $|T_n(x)|=\sqrt n\,|x_n|\le\sqrt n\,\|x\|_2$: $\|T_n\|\le\sqrt n$. Con $x=e_n$ (il vettore con $1$ in posizione $n$, norma $1$): $T_n(e_n)=\sqrt n$, quindi $\|T_n\|\ge\sqrt n$. Combinando, $\|T_n\|=\sqrt n$ esattamente: **le norme crescono senza limite**, $\sup_n\|T_n\|=\infty$.

**Passo 2 — applichiamo il contronominale di Banach-Steinhaus.** Se la famiglia $\{T_n\}$ fosse puntualmente limitata (cioè $\sup_n|T_n(x)|<\infty$ per **ogni** $x\in\ell^2$ fissato), il teorema (essendo $\ell^2$ di Banach) garantirebbe $\sup_n\|T_n\|<\infty$ — ma il passo 1 mostra che questo è falso. Dunque la famiglia **non può essere** puntualmente limitata: esiste (almeno) un $x\in\ell^2$ con $\sup_n|T_n(x)|=\infty$. Il teorema garantisce l'esistenza di un tale $x$ senza indicarlo — costruiamone uno esplicito.

**Passo 3 — costruzione esplicita di $x$.** Definiamo $x_n=1/k$ se $n=2^k$ per qualche intero $k\ge1$, e $x_n=0$ altrimenti (una successione concentrata su un sottoinsieme molto rado di indici, che cresce esponenzialmente).

**Passo 4 — verifichiamo $x\in\ell^2$.** $\sum_nx_n^2=\sum_{k=1}^\infty\dfrac1{k^2}=\dfrac{\pi^2}6<\infty$ (serie di Basilea, notebook 4.10): $x\in\ell^2$.

**Passo 5 — verifichiamo che $T_n(x)$ è illimitata.** Per $n=2^k$: $T_{2^k}(x)=\sqrt{2^k}\cdot x_{2^k}=\sqrt{2^k}/k=2^{k/2}/k$. Essendo l'esponenziale $2^{k/2}$ a crescita più rapida di qualunque potenza di $k$ (fatto standard sui limiti, notebook 4.2/10.10), $2^{k/2}/k\to\infty$ quando $k\to\infty$. Dunque $\sup_nT_n(x)\ge\sup_k2^{k/2}/k=\infty$.

**Conclusione:** $x=(x_n)$ con $x_{2^k}=1/k$ (e $0$ altrove) sta in $\ell^2$ ma ha $\sup_n|T_n(x)|=\infty$: l'esempio esplicito la cui esistenza era garantita, non costruttivamente, dal contronominale di Banach-Steinhaus.

```python
import numpy as np

# Verifica numerica di ||T_n|| = sqrt(n): applicando T_n al vettore e_n (base canonica)
def T_n(x, n):
    if n <= len(x):
        return np.sqrt(n) * x[n-1]
    return 0

for n in [1, 4, 9, 100]:
    e_n = np.zeros(max(n, 10))
    e_n[n-1] = 1
    valore = T_n(e_n, n)
    print(f"T_{n}(e_{n}) = {valore} = sqrt({n}) = {np.sqrt(n):.4f}")

print("\n||T_n|| = sqrt(n): cresce senza limite -> sup_n ||T_n|| = infinito")
print("Per Banach-Steinhaus (contronominale), deve esistere x in ell^2 con sup_n |T_n(x)| = infinito")
print("(la costruzione esplicita di tale x richiede un argomento piu' fine, es. una serie a blocchi)")
```

## Esercizi da risolvere

1. Dimostra la parte facile di Banach-Steinhaus: se $\sup_i\|T_i\|=C<\infty$ (limitatezza uniforme), allora per ogni $v$, $\sup_i\|T_iv\|\le C\|v\|<\infty$ (limitatezza puntuale) — cioè la direzione opposta a quella dimostrata nella teoria, banale ma da giustificare con precisione.
2. Su $\mathbb R^n$ (dimensione finita, sempre di Banach), spiega perché il principio della limitatezza uniforme è "automatico" per una famiglia **finita** di operatori $T_1,\ldots,T_k$, senza bisogno del teorema di Baire (suggerimento: usa che il massimo di un numero finito di numeri finiti è finito).
3. Verifica esplicitamente che $\|T_n\|=\sqrt n$ per l'operatore $T_n(x)=\sqrt n\,x_n$ su $\ell^2$ dell'esercizio svolto, mostrando sia la maggiorazione (Cauchy-Schwarz) sia che è raggiunta (con $x=e_n$).
4. *(più impegnativo)* Dimostra che se $x\in\ell^2$ soddisfa $|x_n|\ge c/\sqrt n$ per **infiniti** indici $n$ (con $c>0$ fissato), allora $\sup_n\sqrt n\,|x_n|\ge c$; usa questo per spiegare perché **nessuna** successione della forma $x_n=c/\sqrt n$ (per ogni $n$, non solo infiniti) può stare in $\ell^2$ — verificando così che il controesempio dell'esercizio svolto deve necessariamente essere concentrato solo su un sottoinsieme rado di indici, non su una formula uniforme in $n$.

**1.** Per ogni $v\in V$ e ogni $i\in I$: $\|T_iv\|\le\|T_i\|\|v\|\le C\|v\|$ (definizione di norma dell'operatore, notebook 14.5, poi limitatezza uniforme $\|T_i\|\le C$). Dunque $\sup_i\|T_iv\|\le C\|v\|$, un numero finito (essendo $C,\|v\|$ entrambi finiti): la famiglia è puntualmente limitata in ogni $v$.

**2.** Per una famiglia finita $T_1,\ldots,T_k$, ciascun $\|T_j\|$ è un numero finito (essendo ogni $T_j$ un singolo operatore limitato, notebook 14.5). Il massimo di un numero **finito** di numeri finiti, $\max(\|T_1\|,\ldots,\|T_k\|)$, è esso stesso finito (banale, non serve alcun argomento di completezza o di Baire): questo massimo è automaticamente un limite uniforme per tutta la famiglia finita. Il teorema di Banach-Steinhaus diventa interessante solo per famiglie **infinite**, dove "il massimo" potrebbe non esistere a priori.

**3.** *Maggiorazione*: per $x\in\ell^2$, $|T_n(x)|=\sqrt n|x_n|$. Essendo $x_n^2\le\sum_mx_m^2=\|x\|_2^2$ (un singolo termine non negativo è al più la somma di tutti, essendo gli altri termini $\ge0$), $|x_n|\le\|x\|_2$, quindi $|T_n(x)|\le\sqrt n\,\|x\|_2$: $\|T_n\|\le\sqrt n$. *Raggiungimento*: con $x=e_n$ (norma $\|e_n\|_2=1$), $T_n(e_n)=\sqrt n\cdot1=\sqrt n$, quindi $|T_n(e_n)|/\|e_n\|_2=\sqrt n/1=\sqrt n$: l'estremo superiore nella definizione di $\|T_n\|$ raggiunge $\sqrt n$, dando $\|T_n\|\ge\sqrt n$. Combinando, $\|T_n\|=\sqrt n$.

**4.** Se $|x_n|\ge c/\sqrt n$ per un particolare $n$, allora $\sqrt n\,|x_n|\ge\sqrt n\cdot c/\sqrt n=c$: dunque $\sup_n\sqrt n\,|x_n|\ge c$ (basta che questo accada per uno solo degli infiniti indici dell'ipotesi, perché l'estremo superiore su tutti gli $n$ sia almeno $c$). Applicando questo con $x_n=c/\sqrt n$ per **ogni** $n$ (non solo infiniti): se una tale successione stesse in $\ell^2$, dovrebbe essere $\sum_nx_n^2=c^2\sum_n1/n<\infty$ — ma $\sum_n1/n$ è la serie armonica, **divergente** (notebook 4.9). Dunque nessuna successione $x_n=c/\sqrt n$ (con $c>0$ fissato, su tutti gli indici) appartiene a $\ell^2$: un controesempio a Banach-Steinhaus per questa famiglia non può avere una formula "uniforme" del tipo $1/\sqrt n$ su tutti gli indici, e deve invece concentrarsi solo su un sottoinsieme raro (come $n=2^k$ nell'esercizio svolto), dove la crescita di $\sqrt n\,|x_n|$ resta compatibile con la sommabilità di $x_n^2$ richiesta da $\ell^2$.

## Riepilogo finale

- Il **teorema di categoria di Baire** dice che uno spazio metrico completo non può essere unione numerabile di chiusi "sottili": lo strumento tecnico dietro i tre grandi teoremi di questa parte del modulo.
- Il **principio della limitatezza uniforme** (Banach-Steinhaus) trasforma la limitatezza **puntuale** di una famiglia di operatori (su uno spazio di Banach) in limitatezza **uniforme** delle loro norme — un salto qualitativo non banale, reso possibile solo dalla completezza.
- La completezza del dominio è **essenziale**: il teorema fallisce su spazi normati incompleti.
- Applicazione classica: l'esistenza (non costruttiva) di funzioni continue con serie di Fourier divergente in un punto, dedotta dal contronominale del teorema applicato agli operatori di somma parziale.
