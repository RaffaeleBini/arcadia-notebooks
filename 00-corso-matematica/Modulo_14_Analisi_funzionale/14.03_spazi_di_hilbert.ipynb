# 14.3 — Spazi di Hilbert

**Prerequisiti:** 14.2 Spazi di Banach, 6.20 Prodotti scalari, 6.21 Ortogonalità, 6.22 Basi ortonormali.


## Perché ci interessa

Uno spazio di Banach (notebook 14.2) ha una norma, ma non necessariamente un **prodotto scalare**: in $\mathbb R^n$ (notebook 6.20), il prodotto scalare non solo induce la norma euclidea, ma fornisce anche l'**angolo** tra vettori, la nozione di **ortogonalità** (notebook 6.21) e le **proiezioni** su sottospazi. Uno **spazio di Hilbert** è uno spazio di Banach la cui norma proviene da un prodotto scalare: la struttura più ricca e "geometrica" tra quelle infinito-dimensionali, dove concetti come basi ortonormali (notebook 6.22) e il teorema di Pitagora si estendono naturalmente, con applicazioni dirette alla meccanica quantistica e all'elaborazione dei segnali (serie di Fourier).

## Teoria, passo dopo passo

### Prodotto scalare (o interno) su uno spazio vettoriale

Un **prodotto scalare** su uno spazio vettoriale $V$ (reale) è una funzione $\langle\cdot,\cdot\rangle:V\times V\to\mathbb R$ tale che, per ogni $u,v,w\in V$ e scalare $\alpha$: (1) $\langle v,v\rangle\ge0$, con uguaglianza solo per $v=0$; (2) $\langle u,v\rangle=\langle v,u\rangle$ (simmetria); (3) $\langle\alpha u+w,v\rangle=\alpha\langle u,v\rangle+\langle w,v\rangle$ (linearità nel primo argomento). Questi sono esattamente gli assiomi verificati per il prodotto scalare euclideo nel notebook 6.20, ora astratti.

### Il prodotto scalare induce una norma: la disuguaglianza di Cauchy-Schwarz generalizzata

Ponendo $\|v\|=\sqrt{\langle v,v\rangle}$, si verifica (esattamente come nel notebook 6.20) la **disuguaglianza di Cauchy-Schwarz**: $|\langle u,v\rangle|\le\|u\|\,\|v\|$, da cui segue la disuguaglianza triangolare per $\|\cdot\|$ (stessa dimostrazione, notebook 6.20): $\|\cdot\|$ è una vera norma (notebook 14.1).

### Definizione di spazio di Hilbert

Uno **spazio di Hilbert** è uno spazio vettoriale $H$ con prodotto scalare $\langle\cdot,\cdot\rangle$, **completo** rispetto alla norma indotta $\|v\|=\sqrt{\langle v,v\rangle}$ (cioè, uno spazio di Banach, notebook 14.2, la cui norma proviene da un prodotto scalare).

**Esempio fondamentale: $L^2(X,\mu)$.** Definendo $\langle f,g\rangle=\int fg\,d\mu$ (notebook 13.4), si verifica che $\langle f,f\rangle=\int f^2\,d\mu=\|f\|_2^2$ (notazione del notebook 13.8): la norma $L^2$ **proviene** da questo prodotto scalare. Essendo $L^2$ completo (Riesz-Fischer, notebook 13.8), $L^2$ è uno spazio di Hilbert — l'**unico** tra gli spazi $L^p$ ad avere questa struttura extra (per $p\ne2$, $\|\cdot\|_p$ non deriva da alcun prodotto scalare, un fatto collegato all'esercizio sotto sulla legge del parallelogramma).

### La legge del parallelogramma: cosa distingue Hilbert da Banach generico

**Teorema.** In uno spazio di Hilbert, $\|u+v\|^2+\|u-v\|^2=2\|u\|^2+2\|v\|^2$ per ogni $u,v$.

**Dimostrazione.** Espandendo con la bilinearità del prodotto scalare: $\|u+v\|^2=\langle u+v,u+v\rangle=\langle u,u\rangle+2\langle u,v\rangle+\langle v,v\rangle=\|u\|^2+2\langle u,v\rangle+\|v\|^2$; analogamente $\|u-v\|^2=\|u\|^2-2\langle u,v\rangle+\|v\|^2$. Sommando, i termini $2\langle u,v\rangle$ si cancellano: $\|u+v\|^2+\|u-v\|^2=2\|u\|^2+2\|v\|^2$. $\blacksquare$ Geometricamente: in un parallelogramma, la somma dei quadrati delle diagonali è il doppio della somma dei quadrati dei lati — un fatto elementare di geometria euclidea (collegabile al notebook 2.x) che qui vale in **ogni** dimensione, anche infinita.

**Uso per escludere il prodotto scalare in $L^p$, $p\ne2$.** Si dimostra (esercizio) che la norma $\|\cdot\|_\infty$ su $C([a,b])$ **non** soddisfa la legge del parallelogramma: dunque, per il teorema (contronominale), $\|\cdot\|_\infty$ non può provenire da nessun prodotto scalare — $C([a,b])$ con la norma del sup è di Banach ma non di Hilbert.

### Ortogonalità e proiezione ortogonale

Due elementi $u,v\in H$ sono **ortogonali** ($u\perp v$) se $\langle u,v\rangle=0$ (notebook 6.21). **Teorema di Pitagora**: se $u\perp v$, $\|u+v\|^2=\|u\|^2+\|v\|^2$ (segue immediatamente dall'espansione sopra, con $\langle u,v\rangle=0$).

**Teorema della proiezione (enunciato).** Se $M\subseteq H$ è un sottospazio **chiuso**, ogni $v\in H$ si scrive in modo **unico** come $v=m+n$ con $m\in M$ e $n\perp M$ (cioè $n$ ortogonale a ogni elemento di $M$). L'elemento $m$, detto **proiezione ortogonale** di $v$ su $M$, è il punto di $M$ più vicino a $v$: $\|v-m\|=\min_{w\in M}\|v-w\|$. Questo teorema generalizza la proiezione ortogonale di un vettore su un sottospazio del notebook 6.21 (dove $M$ era di dimensione finita) a sottospazi chiusi di dimensione anche infinita — la completezza di $H$ (che manca in generale a uno spazio con solo prodotto scalare, non completo) è essenziale per la dimostrazione, che usa un argomento di minimizzazione basato sulla legge del parallelogramma per costruire una successione di Cauchy che approssima il minimo.

```python
import numpy as np

# Verifica numerica della legge del parallelogramma in L^2 (approssimato discretamente)
np.random.seed(0)
n_punti = 1000
x = np.linspace(0, 1, n_punti)

f = np.sin(2*np.pi*x)
g = np.cos(3*np.pi*x)

def prodotto_scalare_L2(f, g, x):
    return np.trapz(f*g, x)

def norma_L2(f, x):
    return np.sqrt(prodotto_scalare_L2(f, f, x))

lhs = norma_L2(f+g, x)**2 + norma_L2(f-g, x)**2
rhs = 2*norma_L2(f, x)**2 + 2*norma_L2(g, x)**2
print(f"||f+g||^2 + ||f-g||^2 = {lhs:.5f}")
print(f"2||f||^2 + 2||g||^2 = {rhs:.5f}")
print(f"Legge del parallelogramma verificata: {np.isclose(lhs, rhs, rtol=1e-2)}")

# Verifica che la norma del sup NON soddisfa il parallelogramma
def norma_sup(f):
    return np.max(np.abs(f))

lhs_sup = norma_sup(f+g)**2 + norma_sup(f-g)**2
rhs_sup = 2*norma_sup(f)**2 + 2*norma_sup(g)**2
print(f"\nCon norma del sup: ||f+g||_inf^2+||f-g||_inf^2 = {lhs_sup:.5f}, 2(||f||_inf^2+||g||_inf^2) = {rhs_sup:.5f}")
print(f"Coincidono (dovrebbero fallire in generale)? {np.isclose(lhs_sup, rhs_sup, rtol=1e-2)}")
```

## Esercizio svolto e commentato

**Problema:** dimostra che $\|\cdot\|_\infty$ su $\mathbb R^2$ non soddisfa la legge del parallelogramma, esibendo un controesempio esplicito con $u=(1,0)$, $v=(0,1)$.

**Ragionamento:**

**Passo 1 — calcoliamo $\|u+v\|_\infty$ e $\|u-v\|_\infty$.** $u+v=(1,1)$: $\|u+v\|_\infty=\max(1,1)=1$. $u-v=(1,-1)$: $\|u-v\|_\infty=\max(1,1)=1$. Dunque $\|u+v\|_\infty^2+\|u-v\|_\infty^2=1+1=2$.

**Passo 2 — calcoliamo $\|u\|_\infty$ e $\|v\|_\infty$.** $\|u\|_\infty=\max(1,0)=1$. $\|v\|_\infty=\max(0,1)=1$. Dunque $2\|u\|_\infty^2+2\|v\|_\infty^2=2\cdot1+2\cdot1=4$.

**Passo 3 — confrontiamo.** $\|u+v\|_\infty^2+\|u-v\|_\infty^2=2\ne4=2\|u\|_\infty^2+2\|v\|_\infty^2$. La legge del parallelogramma **fallisce**.

**Conclusione:** essendo $(\mathbb R^2,\|\cdot\|_\infty)$ uno spazio dove la legge del parallelogramma non vale, per il contronominale del teorema della teoria, $\|\cdot\|_\infty$ **non può provenire da nessun prodotto scalare** — $(\mathbb R^2,\|\cdot\|_\infty)$ è uno spazio normato completo (di Banach, essendo dimensione finita, notebook 14.2) ma **non** è uno spazio di Hilbert. **Punto metodologico:** questo esempio, per quanto elementare (dimensione finita), rivela lo stesso fenomeno strutturale della norma $L^\infty$ o $C([a,b])$ sugli spazi di funzioni: la legge del parallelogramma è un test **algebrico rapido** per escludere l'esistenza di un prodotto scalare compatibile, senza dover verificare direttamente l'assenza di ogni possibile candidato.

```python
import numpy as np

u = np.array([1, 0])
v = np.array([0, 1])

def norma_inf(x):
    return np.max(np.abs(x))

lhs = norma_inf(u+v)**2 + norma_inf(u-v)**2
rhs = 2*norma_inf(u)**2 + 2*norma_inf(v)**2
print(f"||u+v||_inf^2 + ||u-v||_inf^2 = {lhs}")
print(f"2||u||_inf^2 + 2||v||_inf^2 = {rhs}")
print(f"Legge del parallelogramma fallisce: {lhs} != {rhs} -> ||.||_inf non proviene da un prodotto scalare")

# Confronto con la norma euclidea, dove la legge VALE
def norma_2(x):
    return np.linalg.norm(x)

lhs2 = norma_2(u+v)**2 + norma_2(u-v)**2
rhs2 = 2*norma_2(u)**2 + 2*norma_2(v)**2
print(f"\nCon norma euclidea: {lhs2} == {rhs2}? {np.isclose(lhs2, rhs2)}")
```

## Esercizi da risolvere

1. Dimostra il **teorema di Pitagora generalizzato**: se $v_1,\ldots,v_n$ sono a due a due ortogonali, $\|v_1+\cdots+v_n\|^2=\|v_1\|^2+\cdots+\|v_n\|^2$ (suggerimento: usa l'induzione, notebook 10.4, applicando il caso $n=2$ della teoria a ogni passo).
2. Dimostra l'**identità di polarizzazione**: $\langle u,v\rangle=\dfrac14(\|u+v\|^2-\|u-v\|^2)$ (suggerimento: espandi entrambi i quadrati con la bilinearità, come nella dimostrazione della legge del parallelogramma).
3. Verifica che $\ell^2$ (successioni $x=(x_n)$ con $\sum_nx_n^2<\infty$, l'analogo discreto di $L^2$) con $\langle x,y\rangle=\sum_nx_ny_n$ soddisfa gli assiomi di prodotto scalare (verifica solo simmetria e linearità; la positività richiede la convergenza della serie, che diamo per assunta dalla definizione dello spazio).
4. *(più impegnativo)* Usa la disuguaglianza di Cauchy-Schwarz generalizzata per dimostrare che, se $\sum_nx_n^2<\infty$ e $\sum_ny_n^2<\infty$, allora $\sum_n|x_ny_n|<\infty$ (cioè il prodotto scalare $\langle x,y\rangle=\sum_nx_ny_n$ è ben definito, con la serie assolutamente convergente, per ogni $x,y\in\ell^2$).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** *Base ($n=2$)*: è esattamente il teorema di Pitagora della teoria. *Passo induttivo*: supponiamo vero per $n-1$ vettori a due a due ortogonali, cioè $\|v_1+\cdots+v_{n-1}\|^2=\sum_{i=1}^{n-1}\|v_i\|^2$. Poniamo $w=v_1+\cdots+v_{n-1}$: $w\perp v_n$, perché $\langle w,v_n\rangle=\sum_{i=1}^{n-1}\langle v_i,v_n\rangle=0$ (ogni termine è nullo, essendo i $v_i$ a due a due ortogonali, per linearità del prodotto scalare nel primo argomento). Applicando il caso base a $w$ e $v_n$: $\|w+v_n\|^2=\|w\|^2+\|v_n\|^2=\sum_{i=1}^{n-1}\|v_i\|^2+\|v_n\|^2=\sum_{i=1}^n\|v_i\|^2$. Essendo $w+v_n=v_1+\cdots+v_n$, questo completa l'induzione.

**2.** Dalla dimostrazione della legge del parallelogramma: $\|u+v\|^2=\|u\|^2+2\langle u,v\rangle+\|v\|^2$ e $\|u-v\|^2=\|u\|^2-2\langle u,v\rangle+\|v\|^2$. Sottraendo (invece di sommare, come nella legge del parallelogramma): $\|u+v\|^2-\|u-v\|^2=4\langle u,v\rangle$, da cui $\langle u,v\rangle=\dfrac14(\|u+v\|^2-\|u-v\|^2)$.

**3.** *Simmetria*: $\langle x,y\rangle=\sum_nx_ny_n=\sum_ny_nx_n=\langle y,x\rangle$ (commutatività della moltiplicazione reale, notebook 0.5, termine per termine). *Linearità nel primo argomento*: $\langle\alpha x+z,y\rangle=\sum_n(\alpha x_n+z_n)y_n=\alpha\sum_nx_ny_n+\sum_nz_ny_n=\alpha\langle x,y\rangle+\langle z,y\rangle$ (distributiva, applicata termine per termine, poi linearità della somma di una serie convergente, notebook 4.7).

**4.** Per Cauchy-Schwarz generalizzata (teoria), applicata formalmente come limite di somme parziali finite: per ogni $N$, $\sum_{n=1}^N|x_ny_n|\le\Big(\sum_{n=1}^Nx_n^2\Big)^{1/2}\Big(\sum_{n=1}^Ny_n^2\Big)^{1/2}\le\Big(\sum_{n=1}^\infty x_n^2\Big)^{1/2}\Big(\sum_{n=1}^\infty y_n^2\Big)^{1/2}$ (le somme parziali sono maggiorate dalla somma totale, essendo tutti i termini non negativi). Il membro destro è **finito** per ipotesi ($x,y\in\ell^2$), e **non dipende da $N$**: dunque le somme parziali $\sum_{n=1}^N|x_ny_n|$ sono una successione crescente e limitata superiormente, quindi convergente (teorema delle successioni monotone limitate, notebook 4.4/10.10), cioè $\sum_n|x_ny_n|<\infty$.

## Riepilogo finale

- Uno **spazio di Hilbert** è uno spazio di Banach la cui norma proviene da un **prodotto scalare**: la struttura infinito-dimensionale più vicina alla geometria euclidea familiare.
- La **legge del parallelogramma** ($\|u+v\|^2+\|u-v\|^2=2\|u\|^2+2\|v\|^2$) caratterizza algebricamente quali norme derivano da un prodotto scalare: $L^2$ la soddisfa, $L^p$ per $p\ne2$ e $C([a,b])$ con $\|\cdot\|_\infty$ no.
- Il **teorema di Pitagora** e l'**identità di polarizzazione** collegano norma e prodotto scalare in modo esplicito e invertibile.
- Il **teorema della proiezione ortogonale** su sottospazi chiusi generalizza la proiezione del notebook 6.21 a dimensione infinita, sfruttando in modo essenziale la completezza dello spazio.
