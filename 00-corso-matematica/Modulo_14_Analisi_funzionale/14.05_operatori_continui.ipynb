# 14.5 — Operatori continui

**Prerequisiti:** 14.4 Operatori lineari, 12.4 Continuità (topologica), 14.1 Spazi normati.


## Perché ci interessa

Nel notebook 6.13, ogni applicazione lineare tra spazi di dimensione finita era automaticamente continua: un fatto talmente scontato che non veniva nemmeno menzionato. In dimensione infinita questo **non è più vero**: esistono operatori lineari discontinui, come vedremo con l'operatore di derivazione. Questo notebook identifica la condizione precisa — la **limitatezza** — che rende un operatore lineare continuo, dimostrando che le due nozioni (limitatezza algebrica e continuità topologica) coincidono per gli operatori lineari, un teorema sorprendentemente pulito che semplifica drasticamente la verifica della continuità in analisi funzionale.

## Teoria, passo dopo passo

### Definizione di operatore limitato

Un operatore lineare $T:V\to W$ (tra spazi normati) è **limitato** se esiste $C\ge0$ tale che $\|Tv\|\le C\|v\|$ per ogni $v\in V$. Il più piccolo $C$ che funziona si chiama **norma dell'operatore**: $\|T\|=\sup_{v\ne0}\dfrac{\|Tv\|}{\|v\|}=\sup_{\|v\|=1}\|Tv\|$ (le due espressioni coincidono per omogeneità della norma, notebook 14.1: $\|Tv\|/\|v\|=\|T(v/\|v\|)\|$, e $v/\|v\|$ ha norma $1$).

### Teorema: per operatori lineari, continuità equivale a limitatezza

**Teorema.** Un operatore lineare $T:V\to W$ è continuo (rispetto alle topologie indotte dalle norme, notebook 14.1) se e solo se è limitato.

**Dimostrazione.** *(Limitato $\Rightarrow$ continuo)* Sia $\|Tv\|\le C\|v\|$ per ogni $v$. Per ogni $v_0\in V$ e $\varepsilon>0$, scegliendo $\delta=\varepsilon/C$ (se $C>0$; se $C=0$, $T$ è l'operatore nullo, banalmente continuo): $\|v-v_0\|<\delta\Rightarrow\|Tv-Tv_0\|=\|T(v-v_0)\|\le C\|v-v_0\|<C\delta=\varepsilon$ (usando la linearità di $T$ per $Tv-Tv_0=T(v-v_0)$). Questa è esattamente la definizione $\varepsilon$-$\delta$ di continuità in $v_0$ (notebook 12.7), e vale per ogni $v_0$: $T$ è continuo (ovunque, non solo in un punto — un fatto specifico degli operatori lineari, dove la continuità in un punto si propaga a tutto lo spazio, per la stessa struttura di traslazione appena usata).

*(Continuo $\Rightarrow$ limitato, contronominale)* Supponiamo $T$ **non** limitato: per ogni $n$, esiste $v_n$ con $\|Tv_n\|>n\|v_n\|$ (nessun $C$ funziona, in particolare non $C=n$). Necessariamente $v_n\ne0$; poniamo $w_n=v_n/(n\|v_n\|)$, così $\|w_n\|=1/n\to0$ (cioè $w_n\to0$ in norma). Ma $\|Tw_n\|=\|Tv_n\|/(n\|v_n\|)>n\|v_n\|/(n\|v_n\|)=1$ per ogni $n$: $Tw_n$ non tende a $0=T(0)$, nonostante $w_n\to0$. Dunque $T$ non è continuo in $0$ (una successione che converge a $0$ non ha immagine che converge a $T(0)=0$): $T$ non è continuo. $\blacksquare$

**Perché questo teorema è potente.** Riduce la verifica di un fatto **topologico** (continuità, che richiederebbe di controllare ogni intorno) a un fatto **algebrico/quantitativo** (trovare una singola costante $C$): in pratica, per dimostrare che un operatore è continuo, si cerca sempre una stima $\|Tv\|\le C\|v\|$, mai un argomento $\varepsilon$-$\delta$ diretto.

### Un operatore lineare discontinuo: la derivazione su $C([0,1])$

Consideriamo $D:C^1([0,1])\to C([0,1])$ con la norma $\|\cdot\|_\infty$ su **entrambi** gli spazi (invece che sulla topologia più fine tipicamente usata per $C^1$). Prendiamo $f_n(x)=\sin(n\pi x)/n$: $\|f_n\|_\infty=1/n\to0$, ma $Df_n(x)=\pi\cos(n\pi x)$, quindi $\|Df_n\|_\infty=\pi$ per ogni $n$ — **non tende a $0$**. Per il teorema (contronominale), $D$ **non è limitato**, quindi **non è continuo** rispetto a $\|\cdot\|_\infty$. Questo è il primo esempio concreto, in questo modulo, di un operatore lineare discontinuo — un fenomeno impossibile in dimensione finita (notebook 6.13), dove ogni applicazione lineare è automaticamente continua.

### Lo spazio $\mathcal L(V,W)$ degli operatori limitati è normato (ed è di Banach se $W$ lo è)

**Teorema (enunciato).** $\|T\|$ (definita sopra) è una norma su $\mathcal L(V,W)$ (lo spazio degli operatori **limitati**, un sottospazio di $\mathcal L(V,W)$ del notebook 14.4). Se $W$ è uno spazio di Banach (notebook 14.2), allora $(\mathcal L(V,W),\|\cdot\|)$ è esso stesso uno spazio di Banach — la completezza di $W$ "si trasmette" allo spazio degli operatori, un fatto che useremo nel notebook 14.8.

```python
import numpy as np
import matplotlib.pyplot as plt

def f_n(x, n):
    return np.sin(n*np.pi*x) / n

def Df_n(x, n):
    return np.pi * np.cos(n*np.pi*x)

x = np.linspace(0, 1, 500)
fig, axes = plt.subplots(1, 2, figsize=(11,4))
for n in [1, 3, 10, 30]:
    axes[0].plot(x, f_n(x, n), label=f"f_{n}")
    axes[1].plot(x, Df_n(x, n), label=f"D f_{n}")

axes[0].set_title("f_n: ||f_n||_inf = 1/n -> 0")
axes[0].legend()
axes[1].set_title("D f_n: ||D f_n||_inf = pi, NON tende a 0")
axes[1].legend()
plt.tight_layout()
plt.show()

for n in [1, 5, 20, 100]:
    norma_f = np.max(np.abs(f_n(x, n)))
    norma_Df = np.max(np.abs(Df_n(x, n)))
    print(f"n={n}: ||f_n||_inf = {norma_f:.4f}, ||Df_n||_inf = {norma_Df:.4f}")
print("\nD non e' limitato: nessuna costante C soddisfa ||Df_n||<=C||f_n|| per ogni n")
```

## Esercizio svolto e commentato

**Problema:** dimostra che l'operatore di integrazione $I:C([0,1])\to C([0,1])$, $(If)(x)=\int_0^xf(t)\,dt$ (notebook 14.4), è limitato rispetto a $\|\cdot\|_\infty$, e calcola $\|I\|$ (o almeno una stima precisa).

**Ragionamento:**

**Passo 1 — stimiamo $\|If\|_\infty$ in funzione di $\|f\|_\infty$.** Per ogni $x\in[0,1]$: $|(If)(x)|=\Big|\int_0^xf(t)\,dt\Big|\le\int_0^x|f(t)|\,dt\le\int_0^x\|f\|_\infty\,dt=x\|f\|_\infty\le\|f\|_\infty$ (usando $x\le1$ e $|f(t)|\le\|f\|_\infty$ per ogni $t$). Dunque $\|If\|_\infty=\sup_x|(If)(x)|\le\|f\|_\infty$.

**Passo 2 — concludiamo la limitatezza.** Abbiamo trovato $C=1$ che soddisfa $\|If\|_\infty\le C\|f\|_\infty$ per ogni $f$: $I$ è limitato, con $\|I\|\le1$.

**Passo 3 — verifichiamo che $\|I\|=1$ esattamente (non solo $\le1$).** Prendiamo $f=1$ (funzione costante): $\|f\|_\infty=1$, e $(If)(x)=\int_0^x1\,dt=x$, quindi $\|If\|_\infty=\sup_{x\in[0,1]}x=1$. Dunque $\dfrac{\|If\|_\infty}{\|f\|_\infty}=\dfrac11=1$: l'estremo superiore nella definizione di $\|I\|$ è **raggiunto** da questa $f$, quindi $\|I\|\ge1$. Combinando con il passo 2 ($\|I\|\le1$): $\|I\|=1$.

**Conclusione:** $I$ è limitato (quindi continuo, per il teorema della teoria) con $\|I\|=1$ esattamente. **Contrasto con la teoria:** questo esempio, insieme al controesempio della derivazione, mostra che **integrazione e derivazione si comportano in modo opposto** rispetto alla continuità in norma $\|\cdot\|_\infty$: l'integrazione "smussa" (è continua, con norma finita), mentre la derivazione "amplifica le oscillazioni" (può ingrandire arbitrariamente la norma di funzioni piccole, come $f_n$ della teoria) — un fenomeno con conseguenze profonde in analisi (per esempio, nella teoria delle equazioni differenziali, dove gli operatori differenziali richiedono spazi di funzioni più sofisticati, come gli spazi di Sobolev, per essere trattati rigorosamente).

```python
import numpy as np
from scipy import integrate

def f(x, c=1.0):
    return c * np.ones_like(x)

def I_f(x_val, c=1.0):
    return integrate.quad(lambda t: f(np.array([t]), c)[0], 0, x_val)[0]

x = np.linspace(0, 1, 100)
If_values = np.array([I_f(xi) for xi in x])
print(f"||f||_inf = {np.max(np.abs(f(x))):.4f}")
print(f"||I(f)||_inf = {np.max(np.abs(If_values)):.4f}")
print(f"Rapporto ||If||/||f|| = {np.max(np.abs(If_values))/np.max(np.abs(f(x))):.4f} (atteso: 1, il valore di ||I||)")

# Verifica su altre funzioni che il rapporto resta <= 1
for nome, func in [("sin(pi x)", lambda x: np.sin(np.pi*x)), ("x^2", lambda x: x**2), ("cos(2 pi x)", lambda x: np.cos(2*np.pi*x))]:
    fx = func(x)
    Ifx = np.array([integrate.quad(func, 0, xi)[0] for xi in x])
    rapporto = np.max(np.abs(Ifx)) / np.max(np.abs(fx))
    print(f"f={nome}: rapporto ||If||/||f|| = {rapporto:.4f} (<=1)")
```

## Esercizi da risolvere

1. Dimostra che l'operatore identità $\operatorname{id}:V\to V$ è sempre limitato, con $\|\operatorname{id}\|=1$ (per $V\ne\{0\}$).
2. Dimostra che se $T_1,T_2:V\to W$ sono limitati, allora $T_1+T_2$ è limitato con $\|T_1+T_2\|\le\|T_1\|+\|T_2\|$.
3. Dimostra che se $T:V\to W$ e $S:W\to Z$ sono limitati, allora $S\circ T:V\to Z$ è limitato con $\|S\circ T\|\le\|S\|\|T\|$ (suggerimento: applica due volte la definizione di limitatezza).

**1.** $\|\operatorname{id}(v)\|=\|v\|=1\cdot\|v\|$ per ogni $v$: la costante $C=1$ funziona sempre (con uguaglianza), quindi $\operatorname{id}$ è limitato con $\|\operatorname{id}\|\le1$. Prendendo un qualsiasi $v\ne0$ (esiste, essendo $V\ne\{0\}$): $\|\operatorname{id}(v)\|/\|v\|=\|v\|/\|v\|=1$, quindi l'estremo superiore nella definizione di norma è almeno $1$: $\|\operatorname{id}\|\ge1$. Combinando, $\|\operatorname{id}\|=1$.

**2.** $\|(T_1+T_2)(v)\|=\|T_1v+T_2v\|\le\|T_1v\|+\|T_2v\|$ (disuguaglianza triangolare in $W$, notebook 14.1) $\le\|T_1\|\|v\|+\|T_2\|\|v\|=(\|T_1\|+\|T_2\|)\|v\|$ (usando la limitatezza di $T_1$ e $T_2$). Dunque $C=\|T_1\|+\|T_2\|$ funziona per $T_1+T_2$: è limitato, e $\|T_1+T_2\|\le\|T_1\|+\|T_2\|$ (essendo $\|T_1+T_2\|$ il **più piccolo** $C$ che funziona, per definizione di estremo superiore).

**3.** $\|(S\circ T)(v)\|=\|S(Tv)\|\le\|S\|\|Tv\|$ (limitatezza di $S$, applicata al vettore $Tv\in W$) $\le\|S\|\cdot\|T\|\|v\|=(\|S\|\|T\|)\|v\|$ (limitatezza di $T$). Dunque $C=\|S\|\|T\|$ funziona per $S\circ T$: è limitato, con $\|S\circ T\|\le\|S\|\|T\|$.

**4.** Per ogni $x\in[a,b]$: $|g(x)f(x)|\le\|g\|_\infty|f(x)|\le\|g\|_\infty\|f\|_\infty$ (usando $|g(x)|\le\|g\|_\infty$ e poi $|f(x)|\le\|f\|_\infty$). Prendendo l'estremo superiore su $x$: $\|M_gf\|_\infty\le\|g\|_\infty\|f\|_\infty$, quindi $M_g$ è limitato con $\|M_g\|\le\|g\|_\infty$. Per l'uguaglianza: prendendo $f=1$ (costante), $\|f\|_\infty=1$ e $(M_gf)(x)=g(x)\cdot1=g(x)$, quindi $\|M_gf\|_\infty=\|g\|_\infty$. Dunque $\|M_gf\|_\infty/\|f\|_\infty=\|g\|_\infty/1=\|g\|_\infty$: l'estremo superiore nella definizione di $\|M_g\|$ raggiunge (almeno) $\|g\|_\infty$, quindi $\|M_g\|\ge\|g\|_\infty$. Combinando: $\|M_g\|=\|g\|_\infty$.

## Riepilogo finale

- Un operatore lineare è **limitato** se $\|Tv\|\le C\|v\|$ per una costante fissata $C$; il più piccolo $C$ è la **norma dell'operatore** $\|T\|$.
- **Teorema chiave**: per operatori lineari, continuità $\iff$ limitatezza — riducendo la verifica topologica della continuità a una stima algebrica quantitativa.
- La derivazione su $C([0,1])$ con $\|\cdot\|_\infty$ è un esempio concreto di operatore lineare **discontinuo**: un fenomeno impossibile in dimensione finita (notebook 6.13).
- Lo spazio $\mathcal L(V,W)$ degli operatori limitati è normato, con la disuguaglianza $\|T_1+T_2\|\le\|T_1\|+\|T_2\|$ e $\|S\circ T\|\le\|S\|\|T\|$; se $W$ è di Banach, anche $\mathcal L(V,W)$ lo è.
