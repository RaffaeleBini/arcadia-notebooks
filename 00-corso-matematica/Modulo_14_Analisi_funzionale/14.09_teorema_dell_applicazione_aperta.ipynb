# 14.9 — Teorema dell'applicazione aperta

**Prerequisiti:** 14.8 Principio della limitatezza uniforme, 14.4 Operatori lineari, 12.11 Omeomorfismi.


## Perché ci interessa

Nel notebook 14.4 abbiamo visto che un operatore lineare biettivo non è, in dimensione infinita, automaticamente invertibile con inversa continua. Il **teorema dell'applicazione aperta** — il secondo grande teorema basato sul teorema di Baire (notebook 14.8) — chiude parzialmente questa lacuna: se un operatore lineare limitato tra spazi di Banach è **suriettivo**, allora manda aperti in aperti (proprietà che, per un omeomorfismo, notebook 12.11, garantirebbe automaticamente la continuità dell'inversa). La conseguenza pratica più usata è il **teorema dell'inversa limitata**: un operatore lineare biettivo e continuo tra spazi di Banach ha automaticamente inversa continua — un fatto banale in dimensione finita (dove ogni applicazione lineare è rappresentata da una matrice, notebook 6.13/6.14), ma che qui richiede uno strumento potente per essere dimostrato in generale.

## Teoria, passo dopo passo

### Applicazioni aperte

Ricordiamo dal notebook 12.9 (esercizio 3) che una funzione $f$ è **aperta** se l'immagine di ogni aperto è aperta — una proprietà indipendente dalla continuità (che riguarda le controimmagini, non le immagini dirette, notebook 12.4).

### Enunciato del teorema dell'applicazione aperta

**Teorema (dell'applicazione aperta).** Siano $V,W$ spazi di Banach, e $T:V\to W$ un operatore lineare limitato e **suriettivo**. Allora $T$ è un'applicazione aperta: $T(U)$ è aperto in $W$ per ogni aperto $U\subseteq V$.

**Dimostrazione (schema, usando Baire e Banach-Steinhaus in spirito).** Il passo cruciale è mostrare che $T(B_V(0,1))$ (l'immagine della palla unitaria di $V$) contiene una palla $B_W(0,r)$ per qualche $r>0$. Poiché $T$ è suriettivo, $W=\bigcup_nT(B_V(0,n))=\bigcup_nnT(B_V(0,1))$ (usando la linearità: $T(B_V(0,n))=nT(B_V(0,1))$, un riscalamento). Per Baire (essendo $W$ completo), l'insieme $\overline{nT(B_V(0,1))}$ ha interno non vuoto per qualche $n$, quindi anche $\overline{T(B_V(0,1))}$ ha interno non vuoto (riscalando): esiste $B_W(w_0,\varepsilon)\subseteq\overline{T(B_V(0,1))}$. Un argomento di simmetria (usando $-T(v)=T(-v)$, essendo $T$ lineare) mostra che si può prendere $w_0=0$: $B_W(0,\varepsilon)\subseteq\overline{T(B_V(0,1))}$. Il passo tecnico successivo (che omettiamo nei dettagli, ma è dove entra in modo essenziale la **completezza di $V$**, tramite un argomento di serie telescopica costruita per approssimazioni successive) rimuove la chiusura: $B_W(0,\varepsilon/2)\subseteq T(B_V(0,1))$ **senza chiusura**. Da qui, per omogeneità e traslazione, si deduce che $T$ manda ogni palla aperta in un insieme che contiene una palla aperta attorno all'immagine del centro, il che implica che $T$ è aperta su ogni aperto (unione di palle, notebook 12.2). $\blacksquare$

### Il teorema dell'inversa limitata

**Corollario.** Se $T:V\to W$ è un operatore lineare limitato e **biettivo** tra spazi di Banach, allora $T^{-1}:W\to V$ è anch'esso limitato (continuo).

**Dimostrazione.** Essendo $T$ biettivo, in particolare è suriettivo: per il teorema, $T$ è aperta. Per $T^{-1}$ continua, basta (per il notebook 12.4) che la controimmagine di ogni aperto tramite $T^{-1}$ sia aperta — ma la controimmagine di $U$ tramite $T^{-1}$ è esattamente $T(U)$ (per la biettività: $(T^{-1})^{-1}(U)=T(U)$), che è aperto per il teorema. Dunque $T^{-1}$ è continua, quindi limitata (per l'equivalenza del notebook 14.5, applicata a $T^{-1}$). $\blacksquare$

**Perché questo è sorprendente.** In dimensione finita (notebook 6.13), un'applicazione lineare biettiva ha sempre inversa continua — un fatto quasi "gratuito", derivante dalla rappresentazione matriciale (l'inversa di una matrice invertibile è ancora una matrice, quindi automaticamente continua). In dimensione infinita, questo argomento matriciale non esiste più: il teorema dell'applicazione aperta è ciò che **sostituisce** quell'argomento elementare, richiedendo in cambio la completezza di **entrambi** gli spazi.

### Un'applicazione: confronto di norme complete

**Corollario.** Se $\|\cdot\|_a$ e $\|\cdot\|_b$ sono due norme complete (cioè, entrambe rendono $V$ uno spazio di Banach) su uno stesso spazio vettoriale $V$, e $\|v\|_b\le C\|v\|_a$ per ogni $v$ (una sola direzione di confronto), allora le due norme sono **equivalenti** (notebook 14.1).

**Dimostrazione.** L'operatore identità $\operatorname{id}:(V,\|\cdot\|_a)\to(V,\|\cdot\|_b)$ è limitato per ipotesi ($\|\operatorname{id}(v)\|_b=\|v\|_b\le C\|v\|_a$), biettivo (è l'identità), e tra spazi di Banach (per ipotesi). Per il teorema dell'inversa limitata, $\operatorname{id}^{-1}=\operatorname{id}:(V,\|\cdot\|_b)\to(V,\|\cdot\|_a)$ è limitato: esiste $C'$ con $\|v\|_a\le C'\|v\|_b$ per ogni $v$. Combinando le due direzioni: $\|\cdot\|_a$ e $\|\cdot\|_b$ sono equivalenti. $\blacksquare$ Un fatto notevole: **una sola** disuguaglianza tra norme complete implica automaticamente l'altra — un lusso specifico della completezza che non vale per norme generiche.

```python
import numpy as np

# Illustrazione del teorema dell'inversa limitata su un caso finito-dimensionale (dove e' sempre vero)
A = np.array([[2, 1], [1, 3]])  # matrice invertibile, rappresenta un operatore lineare biettivo su R^2
A_inv = np.linalg.inv(A)

print("A =\n", A)
print("A^-1 =\n", A_inv)
print("A * A^-1 =\n", np.round(A @ A_inv, 10))

# Verifica che sia A sia A^-1 sono limitati (le loro norme operatoriali sono finite)
norma_A = np.linalg.norm(A, ord=2)  # norma spettrale (operatoriale indotta da ||.||_2)
norma_A_inv = np.linalg.norm(A_inv, ord=2)
print(f"\n||A|| = {norma_A:.4f}, ||A^-1|| = {norma_A_inv:.4f} (entrambe finite, come garantito)")
print("In dimensione finita questo e' automatico; in dimensione infinita richiede il teorema dell'applicazione aperta")
```

## Esercizio svolto e commentato

**Problema:** su $V=\mathbb R^2$, considera le due norme $\|\cdot\|_1$ (taxicab) e $\|\cdot\|_2$ (euclidea), entrambe complete (essendo $\mathbb R^2$ di dimensione finita, notebook 14.2). Verifica direttamente (senza il teorema dell'applicazione aperta, che qui è "overkill" ma istruttivo da confrontare) che $\|v\|_2\le\|v\|_1$ implica l'esistenza di una costante nell'altra direzione, e confronta questo con l'argomento generale del corollario della teoria.

**Ragionamento:**

**Passo 1 — verifichiamo $\|v\|_2\le\|v\|_1$.** Per $v=(v_1,v_2)$: $\|v\|_2^2=v_1^2+v_2^2\le v_1^2+v_2^2+2|v_1||v_2|=(|v_1|+|v_2|)^2=\|v\|_1^2$ (il termine aggiunto è $\ge0$), quindi $\|v\|_2\le\|v\|_1$ (identico all'esercizio del notebook 12.7 sul confronto tra metriche).

**Passo 2 — applichiamo il corollario della teoria.** Essendo $(\mathbb R^2,\|\cdot\|_1)$ e $(\mathbb R^2,\|\cdot\|_2)$ entrambi di Banach (dimensione finita, notebook 14.2), e valendo $\|v\|_2\le1\cdot\|v\|_1$ (con $C=1$), il corollario garantisce l'esistenza di $C'$ con $\|v\|_1\le C'\|v\|_2$ per ogni $v$ — **senza dover calcolare $C'$ esplicitamente**.

**Passo 3 — confrontiamo con il calcolo diretto (già noto dal notebook 12.7).** Il notebook 12.7 aveva dimostrato direttamente $\|v\|_1\le\sqrt2\,\|v\|_2$ (usando $(a+b)^2\le2(a^2+b^2)$): dunque $C'=\sqrt2$ funziona esplicitamente, coerente con l'esistenza garantita (ma non quantificata) dal corollario.

**Conclusione:** in questo caso a dimensione finita, il calcolo diretto è più informativo (dà la costante esatta $\sqrt2$), mentre il corollario della teoria garantisce solo l'**esistenza** di una qualche costante. **Il vero valore del corollario** emerge in dimensione infinita, dove un calcolo diretto della costante può essere proibitivo o impossibile da ottenere esplicitamente, mentre l'argomento astratto (basato su Baire) garantisce comunque l'equivalenza delle norme, purché entrambe siano complete.

```python
import numpy as np

def norma_1(v):
    return np.sum(np.abs(v))

def norma_2(v):
    return np.linalg.norm(v)

np.random.seed(1)
rapporti = []
for _ in range(1000):
    v = np.random.randn(2)
    r = norma_1(v) / norma_2(v)
    rapporti.append(r)

print(f"Rapporto ||v||_1/||v||_2: min={min(rapporti):.4f}, max={max(rapporti):.4f}")
print(f"Teoria: deve stare tra 1 e sqrt(2)={np.sqrt(2):.4f}")
```

## Riepilogo finale

- Il **teorema dell'applicazione aperta** (Baire, di nuovo) dice che un operatore lineare limitato e suriettivo tra spazi di Banach manda aperti in aperti.
- Il **teorema dell'inversa limitata**, sua conseguenza diretta, garantisce che un operatore lineare biettivo e continuo tra spazi di Banach abbia automaticamente inversa continua — sostituendo, in dimensione infinita, l'argomento matriciale banale della dimensione finita (notebook 6.13/6.14).
- **Corollario pratico**: due norme complete su uno stesso spazio, con una sola disuguaglianza tra loro, sono automaticamente equivalenti — un lusso specifico della completezza.
- Questi risultati mostrano ancora una volta il ruolo centrale della completezza (e del teorema di Baire) nel distinguere la teoria degli spazi di Banach da quella, spesso ingannevolmente simile, dell'algebra lineare a dimensione finita.
