## Perché ci interessa

Nel notebook 14.5 abbiamo studiato operatori $T:V\to W$ tra spazi normati generali. Il caso particolare $W=\mathbb R$ (o $\mathbb C$) — dove l'operatore restituisce un semplice numero anziché un vettore — è così importante da meritare un nome proprio: **funzionale**. I funzionali lineari continui misurano "quanto" un vettore si proietta lungo una direzione fissata, generalizzando $v\mapsto\langle v,u\rangle$ per un $u$ fissato (notebook 6.20). In uno spazio di Hilbert, il **teorema di rappresentazione di Riesz** mostra che **ogni** funzionale continuo ha esattamente questa forma: un risultato elegante che identifica uno spazio di Hilbert con il proprio "spazio dei funzionali".

## Teoria, passo dopo passo

### Definizione di funzionale lineare (continuo)

Un **funzionale lineare** su uno spazio vettoriale $V$ è un operatore lineare $\varphi:V\to\mathbb R$ (caso particolare del notebook 14.4, con $W=\mathbb R$). È **continuo** (equivalentemente, **limitato**, per il teorema del notebook 14.5) se esiste $C$ con $|\varphi(v)|\le C\|v\|$ per ogni $v$. L'insieme dei funzionali lineari continui su $V$, indicato $V^*$ (il **duale topologico** di $V$), è uno spazio normato (caso particolare di $\mathcal L(V,\mathbb R)$, notebook 14.5), con $\|\varphi\|=\sup_{\|v\|=1}|\varphi(v)|$.

### Esempi fondamentali

- Su $\mathbb R^n$: $\varphi(v)=v\cdot u$ per un $u\in\mathbb R^n$ fissato (prodotto scalare, notebook 6.20). Ogni funzionale lineare su $\mathbb R^n$ ha questa forma (un fatto elementare di algebra lineare, conseguenza della rappresentazione matriciale del notebook 6.14: un funzionale è una matrice $1\times n$).
- Su $L^2(X,\mu)$: $\varphi(f)=\int fg\,d\mu=\langle f,g\rangle$ per una $g\in L^2$ fissata (prodotto scalare $L^2$, notebook 14.3). Per Cauchy-Schwarz (notebook 14.3), $|\varphi(f)|\le\|f\|_2\|g\|_2$: $\varphi$ è limitato con $\|\varphi\|\le\|g\|_2$.
- Su $C([a,b])$: $\varphi(f)=f(x_0)$ per un punto $x_0$ fissato (il **funzionale di valutazione**). $|\varphi(f)|=|f(x_0)|\le\|f\|_\infty$: limitato con $\|\varphi\|\le1$ (e si verifica, come nell'esercizio, che vale l'uguaglianza).
- Su $C([a,b])$: $\varphi(f)=\int_a^bf(x)\,dx$ (integrazione come funzionale, distinta dall'operatore di integrazione del notebook 14.4, che restituiva una funzione, non un numero).

### Il teorema di rappresentazione di Riesz

**Teorema (Riesz).** Sia $H$ uno spazio di Hilbert (notebook 14.3) e $\varphi:H\to\mathbb R$ un funzionale lineare continuo. Allora esiste **un unico** $u\in H$ tale che $\varphi(v)=\langle v,u\rangle$ per ogni $v\in H$, e $\|\varphi\|=\|u\|$.

**Dimostrazione (schema, usando la proiezione ortogonale).** Se $\varphi=0$, basta $u=0$. Altrimenti, $\ker\varphi=\{v:\varphi(v)=0\}$ è un sottospazio chiuso (controimmagine di $\{0\}$ chiuso tramite $\varphi$ continuo, notebook 12.4) e **proprio** (essendo $\varphi\ne0$). Per il teorema della proiezione ortogonale (notebook 14.3), esiste $w\ne0$ con $w\perp\ker\varphi$ (un elemento nel complemento ortogonale non banale di $\ker\varphi$, che esiste perché $\ker\varphi$ è un sottospazio proprio chiuso). Per ogni $v\in H$, l'elemento $z=\varphi(v)w-\varphi(w)v$ soddisfa $\varphi(z)=\varphi(v)\varphi(w)-\varphi(w)\varphi(v)=0$ (linearità di $\varphi$), quindi $z\in\ker\varphi$, dunque $z\perp w$: $\langle z,w\rangle=0$, cioè $\varphi(v)\langle w,w\rangle-\varphi(w)\langle v,w\rangle=0$, da cui $\varphi(v)=\dfrac{\varphi(w)}{\|w\|^2}\langle v,w\rangle=\Big\langle v,\dfrac{\varphi(w)}{\|w\|^2}w\Big\rangle$. Poniamo $u=\dfrac{\varphi(w)}{\|w\|^2}w$: questo dà la rappresentazione cercata. L'unicità segue da: se $\langle v,u_1\rangle=\langle v,u_2\rangle$ per ogni $v$, allora $\langle v,u_1-u_2\rangle=0$ per ogni $v$, in particolare per $v=u_1-u_2$, dando $\|u_1-u_2\|^2=0$, cioè $u_1=u_2$. $\blacksquare$

**Significato concettuale.** Il teorema di Riesz dice che $H^*\cong H$ (il duale di uno spazio di Hilbert "è" lo spazio stesso, tramite la corrispondenza $u\leftrightarrow\langle\cdot,u\rangle$): una proprietà molto speciale, che **non** vale per spazi di Banach generici (per esempio, il duale di $L^1$ non è $L^1$, ma $L^\infty$ — un fatto che esula da questo notebook, ma che segnala quanto sia privilegiata la struttura di Hilbert).

```python
import numpy as np

# Verifichiamo il teorema di Riesz su R^n (caso finito-dimensionale, gia' noto dall'algebra lineare)
def phi(v):
    # un funzionale lineare arbitrario su R^3
    return 2*v[0] - 3*v[1] + v[2]

# Per Riesz, deve esistere u tale che phi(v) = v . u per ogni v
u = np.array([2, -3, 1])  # ricostruito dai coefficienti di phi

np.random.seed(0)
for _ in range(5):
    v = np.random.randn(3)
    val_phi = phi(v)
    val_prodotto = np.dot(v, u)
    print(f"v={v.round(3)}: phi(v)={val_phi:.4f}, v.u={val_prodotto:.4f}, coincidono: {np.isclose(val_phi, val_prodotto)}")

# Verifica ||phi|| = ||u|| (norma del funzionale = norma euclidea del vettore rappresentante)
norma_u = np.linalg.norm(u)
print(f"\n||u|| = {norma_u:.4f}")
print("Per Cauchy-Schwarz, |phi(v)| = |v.u| <= ||v|| ||u||, con uguaglianza per v proporzionale a u")
v_test = u / norma_u
print(f"Con v=u/||u||: phi(v) = {phi(v_test):.4f}, atteso ||u|| = {norma_u:.4f}")
```

## Esercizio svolto e commentato

**Problema:** dimostra che il funzionale di valutazione $\varphi_{x_0}(f)=f(x_0)$ su $C([a,b])$ (con $x_0\in[a,b]$ fissato) ha $\|\varphi_{x_0}\|=1$ esattamente.

**Ragionamento:**

**Passo 1 — verifichiamo la linearità.** $\varphi_{x_0}(\alpha f+\beta g)=(\alpha f+\beta g)(x_0)=\alpha f(x_0)+\beta g(x_0)=\alpha\varphi_{x_0}(f)+\beta\varphi_{x_0}(g)$: lineare.

**Passo 2 — stimiamo $|\varphi_{x_0}(f)|$.** $|\varphi_{x_0}(f)|=|f(x_0)|\le\sup_{x\in[a,b]}|f(x)|=\|f\|_\infty$ (banalmente, essendo $x_0$ uno dei punti su cui si prende l'estremo superiore). Dunque $\varphi_{x_0}$ è limitato con $\|\varphi_{x_0}\|\le1$.

**Passo 3 — troviamo una $f$ che raggiunge (quasi) l'uguaglianza.** Prendiamo $f=1$ (funzione costante): $\|f\|_\infty=1$, e $\varphi_{x_0}(f)=f(x_0)=1$. Dunque $|\varphi_{x_0}(f)|/\|f\|_\infty=1/1=1$: l'estremo superiore nella definizione di $\|\varphi_{x_0}\|$ raggiunge $1$, quindi $\|\varphi_{x_0}\|\ge1$.

**Conclusione:** combinando i passi 2 e 3, $\|\varphi_{x_0}\|=1$. **Osservazione concettuale:** questo funzionale **non** è rappresentabile nella forma $\varphi(f)=\int fg\,dx$ per nessuna funzione $g\in L^2([a,b])$ (un fatto che esula dalla dimostrazione qui, ma è collegato al fatto che $C([a,b])$ con $\|\cdot\|_\infty$ non è uno spazio di Hilbert, notebook 14.3): il teorema di Riesz si applica **solo** a spazi di Hilbert, e $\varphi_{x_0}$ mostra un esempio di funzionale continuo "genuinamente diverso" da un prodotto scalare, coerente con l'assenza della struttura hilbertiana in $C([a,b])$.

```python
import numpy as np

def valuta(f, x0):
    return f(x0)

# Verifica numerica su diverse funzioni continue su [0,1], con x0=0.5
x0 = 0.5
funzioni = {
    "f=1 (costante)": lambda x: 1.0,
    "f=x": lambda x: x,
    "f=sin(pi x)": lambda x: np.sin(np.pi * x),
    "f=x^2": lambda x: x**2,
}

x_grid = np.linspace(0, 1, 1000)
for nome, f in funzioni.items():
    norma_f = np.max(np.abs(f(x_grid)))
    valore = valuta(f, x0)
    rapporto = abs(valore) / norma_f if norma_f > 0 else 0
    print(f"{nome}: ||f||_inf = {norma_f:.4f}, f(x0) = {valore:.4f}, rapporto = {rapporto:.4f} (<=1)")

print(f"\nCon f=1: rapporto = 1 esattamente -> ||phi_x0|| = 1")
```

Dunque $\varphi$ è limitato (con $\|\varphi\|\le\|u\|_2=1/\sqrt3$) e rappresentato da $u=(1/2,1/4,1/8,\ldots)$ secondo Riesz.

Per l'assioma di non-degenerazione del prodotto scalare (notebook 14.3, $\langle w,w\rangle=0\Rightarrow w=0$): $u_1-u_2=0$, cioè $u_1=u_2$.

## Riepilogo finale

- Un **funzionale lineare** è un operatore a valori in $\mathbb R$; il suo spazio duale $V^*$ raccoglie tutti i funzionali continui, con la norma indotta.
- Esempi fondamentali: il prodotto scalare con un vettore fissato ($\mathbb R^n$, $L^2$), la valutazione in un punto e l'integrazione (su $C([a,b])$).
- Il **teorema di rappresentazione di Riesz** identifica ogni funzionale continuo su uno spazio di Hilbert con il prodotto scalare rispetto a un unico vettore: $H^*\cong H$, una proprietà speciale degli spazi di Hilbert.
- Il funzionale di valutazione su $C([a,b])$ non è rappresentabile come integrale contro alcuna funzione $L^2$: un segnale che $C([a,b])$ con $\|\cdot\|_\infty$ non è di Hilbert.
