## Perché ci interessa

Nel notebook 6.24 abbiamo dimostrato che ogni matrice simmetrica è diagonalizzabile con una base ortonormale di autovettori — il teorema spettrale in dimensione finita, uno dei risultati più usati dell'algebra lineare. Questo notebook, capstone del Modulo 14 e dell'intero percorso di analisi funzionale, dimostra (enunciandone la struttura con precisione) l'analogo per **operatori compatti autoaggiunti** su uno spazio di Hilbert: pur agendo su uno spazio di dimensione infinita, tali operatori si comportano "quasi come matrici", ammettendo una base ortonormale di autovettori con autovalori che tendono a $0$. Questo teorema è lo strumento che rende rigorosa, per esempio, la teoria delle serie di Fourier come "diagonalizzazione" di operatori differenziali, e apre la porta alla meccanica quantistica, dove gli osservabili fisici sono descritti esattamente da operatori autoaggiunti.

## Teoria, passo dopo passo

### Operatori autoaggiunti

Un operatore $T:H\to H$ su uno spazio di Hilbert è **autoaggiunto** (o simmetrico, notebook 14.10) se $\langle Tv,w\rangle=\langle v,Tw\rangle$ per ogni $v,w\in H$. Per il teorema di Hellinger-Toeplitz (notebook 14.10), ogni operatore autoaggiunto definito su **tutto** $H$ è automaticamente limitato — una precondizione tecnica già acquisita.

### Autovalori di operatori autoaggiunti sono reali; autovettori a autovalori distinti sono ortogonali

**Teorema.** Se $T$ è autoaggiunto e $Tv=\lambda v$ con $v\ne0$, allora $\lambda\in\mathbb R$ (ovvio se lavoriamo già in uno spazio reale, ma il fatto ha senso e si generalizza al caso complesso). Se $Tv_1=\lambda_1v_1$, $Tv_2=\lambda_2v_2$ con $\lambda_1\ne\lambda_2$, allora $v_1\perp v_2$.

**Dimostrazione (seconda parte).** $\lambda_1\langle v_1,v_2\rangle=\langle Tv_1,v_2\rangle=\langle v_1,Tv_2\rangle=\lambda_2\langle v_1,v_2\rangle$ (usando l'autoaggiunzione, poi la definizione di autovettore su entrambi i lati). Dunque $(\lambda_1-\lambda_2)\langle v_1,v_2\rangle=0$; essendo $\lambda_1\ne\lambda_2$, deve essere $\langle v_1,v_2\rangle=0$. $\blacksquare$ Identica, parola per parola, alla dimostrazione del notebook 6.24 per matrici simmetriche.

### Il teorema spettrale per operatori compatti autoaggiunti

**Teorema (spettrale, per operatori compatti autoaggiunti).** Sia $T:H\to H$ compatto (notebook 14.11) e autoaggiunto, su uno spazio di Hilbert $H$. Allora esiste una successione (finita o numerabile) di autovalori reali $\lambda_1,\lambda_2,\ldots$ (eventualmente ripetuti secondo molteplicità, e con $\lambda_n\to0$ se la successione è infinita) e una corrispondente famiglia ortonormale $e_1,e_2,\ldots$ di autovettori, tale che per ogni $v\in H$:
$$Tv=\sum_n\lambda_n\langle v,e_n\rangle e_n.$$

**Perché $\lambda_n\to0$: il ruolo della compattezza.** Se esistessero infiniti autovalori con $|\lambda_n|\ge\varepsilon>0$, i corrispondenti autovettori ortonormali $e_n$ (limitati, $\|e_n\|=1$) darebbero $\|Te_n-Te_m\|^2=\|\lambda_ne_n-\lambda_me_m\|^2=\lambda_n^2+\lambda_m^2\ge2\varepsilon^2$ (per Pitagora, notebook 14.3, essendo $e_n\perp e_m$): una successione $(Te_n)$ senza sottosuccessioni di Cauchy, contraddicendo la compattezza di $T$ (che richiederebbe una sottosuccessione convergente di $(Te_n)$, essendo $(e_n)$ limitata). Dunque gli autovalori "grandi" (in modulo $\ge\varepsilon$, per ogni $\varepsilon$ fissato) possono essere solo in numero **finito**: $\lambda_n\to0$.

**Confronto diretto con il notebook 6.24.** In dimensione finita, la somma nel teorema è **finita** (al più $n=\dim H$ termini) e non c'è bisogno della condizione $\lambda_n\to0$ (irrilevante con finiti termini). In dimensione infinita, la compattezza di $T$ è **esattamente** l'ipotesi che permette alla somma infinita di "comportarsi bene" (convergere in norma), attraverso il decadimento degli autovalori: senza compattezza, il teorema fallisce in generale (un operatore autoaggiunto limitato ma non compatto, come l'identità, può non avere affatto autovettori che formano una base, o avere spettro continuo — un fenomeno che esula da questo notebook introduttivo).

### Un esempio concreto: l'operatore diagonale

Riprendendo l'operatore $T(x_1,x_2,\ldots)=(x_1,x_2/2,x_3/3,\ldots)$ del notebook 14.11 (già dimostrato compatto): è autoaggiunto su $\ell^2$ (verifica diretta: $\langle Tx,y\rangle=\sum_n(x_n/n)y_n=\sum_nx_n(y_n/n)=\langle x,Ty\rangle$), con autovalori $\lambda_n=1/n$ (essendo $Te_n=e_n/n$) e autovettori $e_n$ (la base canonica, già ortonormale). Qui $\lambda_n=1/n\to0$, coerente con il teorema, e la decomposizione spettrale $Tx=\sum_n\dfrac1n\langle x,e_n\rangle e_n=\sum_n\dfrac{x_n}ne_n$ è **esattamente** la definizione di $T$ da cui siamo partiti — un caso in cui il teorema spettrale non aggiunge informazione nuova (l'operatore era già "diagonale in partenza"), ma conferma la coerenza della teoria.

```python
import numpy as np

# Verifica numerica del teorema spettrale sull'operatore diagonale T(x)_n = x_n/n
def T(x):
    n = np.arange(1, len(x)+1)
    return x / n

np.random.seed(0)
x = np.random.randn(10)

# Calcolo diretto
Tx_diretto = T(x)

# Calcolo tramite la formula spettrale: T(x) = sum_n lambda_n <x,e_n> e_n
autovalori = 1 / np.arange(1, 11)
Tx_spettrale = np.zeros(10)
for n in range(10):
    e_n = np.zeros(10)
    e_n[n] = 1
    proiezione = np.dot(x, e_n)
    Tx_spettrale += autovalori[n] * proiezione * e_n

print("T(x) calcolo diretto:   ", Tx_diretto.round(4))
print("T(x) formula spettrale: ", Tx_spettrale.round(4))
print("Coincidono?", np.allclose(Tx_diretto, Tx_spettrale))

# Verifica che gli autovalori tendono a 0
print(f"\nAutovalori lambda_n = 1/n: {autovalori.round(4)}")
print("Tendono a 0, coerentemente con la compattezza di T")
```

## Esercizio svolto e commentato

**Problema:** su $\mathbb R^3$ (caso finito-dimensionale, per collegarsi esplicitamente al notebook 6.24), la matrice $A=\begin{pmatrix}2&1&0\\1&2&0\\0&0&3\end{pmatrix}$ è simmetrica. Trova i suoi autovalori e una base ortonormale di autovettori, e verifica la formula spettrale $Av=\sum_i\lambda_i\langle v,e_i\rangle e_i$ su un vettore specifico.

**Ragionamento:**

**Passo 1 — troviamo gli autovalori.** Il blocco $\begin{pmatrix}2&1\\1&2\end{pmatrix}$ ha autovalori $2\pm1=3,1$ (traccia $4$, determinante $3$, risolvendo $\lambda^2-4\lambda+3=0$, notebook 6.17); il blocco $(3)$ dà l'autovalore $3$. Dunque gli autovalori di $A$ sono $\lambda_1=1,\lambda_2=3,\lambda_3=3$ (con $\lambda_2$ di molteplicità $2$, combinando il contributo del primo blocco e del secondo).

**Passo 2 — troviamo gli autovettori.** Per $\lambda=1$: $(A-I)v=0$ dà $v_1=(1,-1,0)/\sqrt2$ (normalizzato). Per $\lambda=3$ (dal primo blocco): $v_2=(1,1,0)/\sqrt2$. Per $\lambda=3$ (dal secondo blocco): $v_3=(0,0,1)$. Verifica di ortogonalità: $v_1\perp v_2$ (autovalori distinti, $1\ne3$, per il teorema della teoria); $v_3\perp v_1,v_2$ (per costruzione, essendo in blocchi separati); $v_2,v_3$ hanno lo **stesso** autovalore $3$, quindi la loro ortogonalità non è garantita dal teorema (che vale solo per autovalori **distinti**) ma si verifica **direttamente**: $v_2\cdot v_3=0$.

**Passo 3 — verifichiamo la formula spettrale su $v=(1,0,0)$.** $\langle v,v_1\rangle=1/\sqrt2$, $\langle v,v_2\rangle=1/\sqrt2$, $\langle v,v_3\rangle=0$. La formula dà: $\sum_i\lambda_i\langle v,e_i\rangle e_i=1\cdot\dfrac1{\sqrt2}\cdot\dfrac{(1,-1,0)}{\sqrt2}+3\cdot\dfrac1{\sqrt2}\cdot\dfrac{(1,1,0)}{\sqrt2}+3\cdot0\cdot(0,0,1)=\dfrac12(1,-1,0)+\dfrac32(1,1,0)=(2,1,0)$.

**Passo 4 — confrontiamo con il calcolo diretto.** $Av=A(1,0,0)^T=(2,1,0)^T$ (prima colonna di $A$). Coincide con il risultato del passo 3.

**Conclusione:** la formula spettrale riproduce esattamente $Av$, confermando il teorema in questo caso finito-dimensionale — lo stesso enunciato del notebook 6.24, ora visto come **caso particolare** (con somma finita, nessuna condizione su $\lambda_n\to0$) del teorema più generale per operatori compatti autoaggiunti su spazi di Hilbert di dimensione qualsiasi.

```python
import numpy as np

A = np.array([[2, 1, 0], [1, 2, 0], [0, 0, 3]])
autovalori, autovettori = np.linalg.eigh(A)  # eigh usa la simmetria per garantire risultati reali e ortonormali

print("Autovalori:", autovalori)
print("Autovettori (colonne):\n", autovettori.round(4))

v = np.array([1, 0, 0])
Av_diretto = A @ v

Av_spettrale = np.zeros(3)
for i in range(3):
    e_i = autovettori[:, i]
    proiezione = np.dot(v, e_i)
    Av_spettrale += autovalori[i] * proiezione * e_i

print(f"\nA*v (diretto):    {Av_diretto}")
print(f"A*v (spettrale):  {Av_spettrale.round(4)}")
print(f"Coincidono? {np.allclose(Av_diretto, Av_spettrale)}")
```

4. *(più impegnativo)* Usa il teorema spettrale per dimostrare che $\|T\|=\max_n|\lambda_n|$ per un operatore compatto autoaggiunto $T$ (suggerimento: usa la formula spettrale e Pitagora per stimare $\|Tv\|^2=\sum_n\lambda_n^2\langle v,e_n\rangle^2$, poi confrontalo con $\|v\|^2=\sum_n\langle v,e_n\rangle^2$ tramite l'identità di Parseval).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** *Autoaggiunto*: $\langle Tx,y\rangle=\sum_n(x_n/n^2)y_n=\sum_nx_n(y_n/n^2)=\langle x,Ty\rangle$ (stesso argomento dell'esempio della teoria). *Compatto*: identico all'argomento del notebook 14.11 per l'operatore $x_n/n$, approssimando con $T_N$ (troncamento a $N$ termini, rango finito) e stimando $\|T-T_N\|\le1/(N+1)^2\to0$. *Autovalori e autovettori*: $Te_n=e_n/n^2$, quindi $\lambda_n=1/n^2$ con autovettore $e_n$ (base canonica).

**2.** Se l'autospazio per $\lambda\ne0$ avesse dimensione infinita, potremmo estrarre una successione infinita $(v_n)$ ortonormale al suo interno (procedimento di Gram-Schmidt, notebook 6.22, applicato a una base numerabile dell'autospazio). Essendo $Tv_n=\lambda v_n$ per ogni $n$: $\|Tv_n-Tv_m\|^2=\|\lambda v_n-\lambda v_m\|^2=\lambda^2\|v_n-v_m\|^2=\lambda^2(\|v_n\|^2+\|v_m\|^2)=2\lambda^2$ (Pitagora, essendo $v_n\perp v_m$), una costante positiva (essendo $\lambda\ne0$) che non tende a $0$: $(Tv_n)$ non ha sottosuccessioni di Cauchy, contraddicendo la compattezza di $T$ (che richiederebbe una sottosuccessione convergente, essendo $(v_n)$ limitata). Dunque l'autospazio per $\lambda\ne0$ ha dimensione finita.

**3.** Se $T$ non è a rango finito, la sua immagine ha dimensione infinita; per il teorema spettrale, questo richiede infiniti autovalori non nulli $\lambda_n\to0$ (l'immagine è generata dagli autovettori con $\lambda_n\ne0$). Essendo $\lambda_n\to0$, per ogni $\varepsilon>0$ esistono infiniti $n$ con $|\lambda_n|<\varepsilon$, ma **nessun** $\lambda_n$ è mai esattamente $0$ necessariamente — tuttavia, se $T$ non è iniettivo (il caso tipico quando l'immagine ha dimensione infinita strettamente minore dello spazio, o quando $0$ compare come autovalore con autospazio il nucleo), $0$ è autovalore. In ogni caso, per l'esercizio 3 del notebook 14.11, $T$ (essendo compatto su uno spazio di dimensione infinita) non può avere inversa limitata: se la avesse, $\operatorname{id}=T^{-1}T$ sarebbe compatta, assurdo per il teorema di Riesz (notebook 14.11).

**4.** Per la formula spettrale, $Tv=\sum_n\lambda_n\langle v,e_n\rangle e_n$: essendo gli $e_n$ ortonormali, per Pitagora generalizzato (notebook 14.3, esercizio 1) $\|Tv\|^2=\sum_n\lambda_n^2\langle v,e_n\rangle^2$. Per l'identità di Parseval (conseguenza della base ortonormale, collegata al notebook 6.22): $\|v\|^2\ge\sum_n\langle v,e_n\rangle^2$ (uguaglianza se gli $e_n$ formano una base completa di $H$, disuguaglianza di Bessel in generale). Ponendo $M=\max_n|\lambda_n|$ (esiste, essendo $\lambda_n\to0$ quindi limitati, e il massimo di un insieme che include un numero finito di "grandi" valori più una coda che tende a $0$ è raggiunto): $\|Tv\|^2=\sum_n\lambda_n^2\langle v,e_n\rangle^2\le M^2\sum_n\langle v,e_n\rangle^2\le M^2\|v\|^2$, quindi $\|Tv\|\le M\|v\|$, cioè $\|T\|\le M$. Prendendo $v=e_{n_0}$ dove $|\lambda_{n_0}|=M$: $\|Te_{n_0}\|=|\lambda_{n_0}|\|e_{n_0}\|=M$, quindi $\|T\|\ge M$. Combinando, $\|T\|=M=\max_n|\lambda_n|$.

## Riepilogo finale

- Un operatore **autoaggiunto** ha autovalori reali e autovettori a autovalori distinti ortogonali — identico al caso matriciale del notebook 6.24.
- Il **teorema spettrale per operatori compatti autoaggiunti** garantisce una base ortonormale di autovettori con $\lambda_n\to0$: la compattezza è l'ipotesi che sostituisce la finitezza della dimensione, garantendo che la somma spettrale (potenzialmente infinita) si comporti bene.
- Un operatore compatto autoaggiunto su uno spazio di dimensione infinita non è mai invertibile con inversa limitata: la compattezza forza $0$ a essere, in un senso preciso, un punto di accumulazione dello spettro.
- $\|T\|=\max_n|\lambda_n|$: la norma dell'operatore si legge direttamente dal più grande autovalore in modulo, esattamente come per le matrici simmetriche.

Con questo si chiude il Modulo 14: partendo dagli assiomi di norma (notebook 14.1), abbiamo costruito Banach, Hilbert, operatori limitati e i tre grandi teoremi di Baire, per arrivare a un teorema spettrale che **generalizza rigorosamente**, a dimensione infinita, uno dei risultati più usati dell'algebra lineare classica del Modulo 6.
