## Teoria, passo dopo passo

### Ripasso della definizione

Ricordiamo dal notebook 0.19: una funzione $f$ da un insieme $A$ a un insieme $B$, scritta $f:A\to B$, associa a **ogni** elemento $a\in A$ **esattamente un** elemento $f(a)\in B$. L'insieme $A$ si chiama **dominio**, l'insieme $B$ si chiama **codominio** (approfondiremo la distinzione tra codominio e immagine nel notebook 3.4).

### I quattro modi di descrivere una funzione

Una stessa funzione si può presentare in modi diversi, ed è utile riconoscerli tutti:

1. **Formula algebrica**: $f(x)=x^2-3x+2$ (come nei polinomi, notebook 1.1, o le coniche, Modulo 2).
2. **Grafico**: l'insieme dei punti $(x,f(x))$ nel piano cartesiano (notebook 2.1) — il "test della retta verticale" del notebook 0.19 permette di riconoscere se un grafico rappresenta davvero una funzione.
3. **Tabella di valori**: un elenco (necessariamente parziale, se il dominio è infinito) di coppie $(x,f(x))$.
4. **Descrizione a parole o a tratti**: come abbiamo fatto per il valore assoluto (notebook 1.14), definendo la funzione con regole diverse su intervalli diversi.

Sapere passare da una rappresentazione all'altra è un'abilità che useremo continuamente.

### Funzioni come "macchine"

Un'immagine utile: pensa a $f$ come a una macchina che riceve un input $x$ (dal dominio) e restituisce un unico output $f(x)$ (nel codominio), sempre secondo la stessa regola. La stessa macchina, con lo stesso input, restituisce sempre lo stesso output — è esattamente la proprietà di unicità della definizione di funzione (notebook 0.19): non c'è casualità, non c'è ambiguità.

### Uguaglianza tra funzioni

Due funzioni $f$ e $g$ sono **uguali** se e solo se hanno lo stesso dominio e $f(x)=g(x)$ per ogni $x$ nel dominio comune. Questo significa che $f(x)=x$ e $g(x)=\frac{x^2}{x}$ **non sono la stessa funzione**: hanno domini diversi ($f$ è definita per ogni $x$ reale, $g$ richiede $x\ne0$, notebook 1.4) anche se, dove entrambe sono definite, assumono lo stesso valore. Questa distinzione — sottile ma importante — tornerà utile studiando le frazioni algebriche come funzioni (notebook 3.12).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')

# Le quattro rappresentazioni della stessa funzione f(x) = x^2 - 3x + 2

# 1. Formula
f = x**2 - 3*x + 2
print("Formula:", f)

# 2. Grafico
f_numerica = sp.lambdify(x, f, 'numpy')
x_vals = np.linspace(-1, 4, 100)
plt.figure(figsize=(6, 4))
plt.plot(x_vals, f_numerica(x_vals))
plt.axhline(y=0, color='gray', linewidth=0.5)
plt.title("Grafico di f(x) = x² - 3x + 2")
plt.grid(True, alpha=0.3)
plt.show()

# 3. Tabella di valori
print("\nTabella di valori:")
for x_val in range(-1, 5):
    print(f"  f({x_val}) = {f.subs(x, x_val)}")

# Uguaglianza tra funzioni: f(x)=x e g(x)=x^2/x NON sono la stessa funzione (domini diversi)
g = x**2 / x
print("\ng semplificata:", sp.simplify(g))
print("Ma il dominio di g esclude x=0, mentre f(x)=x è definita anche in x=0")
print("g(0) è definita?", g.subs(x, 0))  # darà un errore o un'indeterminazione
````

## Esercizio svolto e commentato

**Problema:** stabilisci se la seguente tabella di valori può rappresentare (una parte di) una funzione, e se sì, indovina una possibile formula.

| $x$ | $-2$ | $-1$ | $0$ | $1$ | $2$ |
|---|---|---|---|---|---|
| $y$ | $4$ | $1$ | $0$ | $1$ | $4$ |

**Ragionamento:**

**Passo 1 — verifichiamo che sia compatibile con una funzione.** Ogni valore di $x$ compare **una sola volta** nella tabella, ciascuno associato a un unico $y$: non c'è nessuna violazione della definizione di funzione (notebook 0.19, unicità dell'immagine). È compatibile.

**Passo 2 — cerchiamo uno schema.** Osserviamo che $y$ sembra dipendere solo dal **valore assoluto** di $x$ (notebook 0.8): $x=-2$ e $x=2$ danno lo stesso $y=4$; $x=-1$ e $x=1$ danno lo stesso $y=1$. Questo suggerisce che $f$ potrebbe essere **pari** (concetto che formalizzeremo nel notebook 3.8): $f(-x)=f(x)$.

**Passo 3 — proviamo un'ipotesi.** Notiamo che $4=2^2$, $1=1^2$, $0=0^2$: sembra che $y=x^2$. Verifichiamo su tutti i punti: $f(-2)=(-2)^2=4$ ✓, $f(-1)=1$ ✓, $f(0)=0$ ✓, $f(1)=1$ ✓, $f(2)=4$ ✓. Tutti coincidono.

**Perché "indovinare" una formula da una tabella non è mai una dimostrazione, solo un'ipotesi plausibile:** una tabella mostra solo un numero **finito** di punti. Infinite funzioni diverse potrebbero passare esattamente per questi cinque punti (per esempio un polinomio di grado più alto costruito apposta). $y=x^2$ è la spiegazione più **semplice** che si adatta ai dati, ma senza altre informazioni (per esempio, sapere che la funzione è un polinomio di grado 2) non possiamo esserne matematicamente certi — un principio che ritroverai spesso quando si modella un fenomeno a partire da dati sperimentali.

````python
import sympy as sp

x = sp.symbols('x')
punti = {-2: 4, -1: 1, 0: 0, 1: 1, 2: 4}

f_ipotesi = x**2
tutti_coincidono = all(f_ipotesi.subs(x, k) == v for k, v in punti.items())
print("L'ipotesi f(x)=x^2 è coerente con tutti i punti della tabella?", tutti_coincidono)
````

## Riepilogo finale

- Una funzione $f:A\to B$ associa a ogni elemento di $A$ (dominio) esattamente un elemento di $B$ (codominio) — la definizione del notebook 0.19, ora al centro di un intero modulo.
- Una funzione si può descrivere con formula, grafico, tabella, o regole a tratti: sono rappresentazioni diverse dello stesso oggetto.
- Due funzioni sono uguali solo se hanno lo **stesso dominio** e assumono lo stesso valore ovunque: espressioni algebricamente equivalenti ma con domini diversi rappresentano funzioni diverse.
- Una tabella di valori finita non determina univocamente una funzione: identificare una formula a partire da pochi punti è un'ipotesi plausibile, non una dimostrazione.
