## Teoria, passo dopo passo

### Riepilogo classificato

Raccogliamo le proprietà di $\sin x$, $\cos x$, $\tan x$ (dal notebook 2.12) nel linguaggio di questo modulo:

| Proprietà | $\sin x$ | $\cos x$ | $\tan x$ |
|---|---|---|---|
| Dominio | $\mathbb{R}$ | $\mathbb{R}$ | $\mathbb{R}\setminus\{\frac\pi2+k\pi\}$ |
| Immagine | $[-1,1]$ | $[-1,1]$ | $\mathbb{R}$ |
| Periodo | $2\pi$ | $2\pi$ | $\pi$ |
| Parità (notebook 3.8) | dispari | pari | dispari |
| Limitata (notebook 3.10) | sì | sì | no |
| Iniettiva su $\mathbb{R}$? | no | no | no (sul dominio) |

Nessuna delle tre è iniettiva sul proprio dominio "naturale": la periodicità stessa lo impedisce (notebook 2.11: infinite soluzioni per ogni equazione elementare, quindi infiniti $x$ con la stessa immagine).

### Monotonia locale del seno

Su un intero periodo $[-\frac\pi2,\frac{3\pi}2]$, il seno alterna tratti di monotonia. Il tratto più naturale da isolare (perché contiene l'origine ed è il più "simmetrico") è:

$$\sin x \text{ è strettamente crescente su } \left[-\frac\pi2,\frac\pi2\right]$$

(da $\sin(-\frac\pi2)=-1$ fino a $\sin(\frac\pi2)=1$: percorre **tutta** l'immagine $[-1,1]$ in questo intervallo). Fuori da questo intervallo, il seno decresce, poi ricomincia a crescere, periodicamente.

### Monotonia locale del coseno

Analogamente, il tratto naturale per il coseno è:

$$\cos x \text{ è strettamente decrescente su } [0,\pi]$$

(da $\cos(0)=1$ fino a $\cos(\pi)=-1$: percorre tutta l'immagine $[-1,1]$).

### Monotonia della tangente

La tangente, a differenza di seno e coseno, è **strettamente crescente su tutto un periodo** (senza inversioni di direzione), per esempio su $\left(-\frac\pi2,\frac\pi2\right)$ — l'intervallo tra due asintoti verticali consecutivi (notebook 2.12). Qui la tangente è già iniettiva su tutto l'intervallo, percorrendo l'intera immagine $\mathbb{R}$ (da $-\infty$ a $+\infty$).

### Perché questi intervalli specifici, e non altri

Per ciascuna funzione, esistono **infiniti** intervalli di monotonia stretta (uno per ogni periodo, per periodicità). La scelta di $[-\frac\pi2,\frac\pi2]$ per il seno, $[0,\pi]$ per il coseno, $\left(-\frac\pi2,\frac\pi2\right)$ per la tangente non è casuale: sono gli intervalli **più vicini all'origine** che contengono un intero "arco" di monotonia e percorrono l'intera immagine. Questa scelta standard è quella su cui si basa la definizione delle funzioni trigonometriche inverse, nel prossimo notebook.

````python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(-2*np.pi, 2*np.pi, 1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(theta, np.sin(theta), color='lightgray', label='sin(θ) completo')
theta_crescente = np.linspace(-np.pi/2, np.pi/2, 100)
axes[0].plot(theta_crescente, np.sin(theta_crescente), color='tab:blue', linewidth=3, label='tratto crescente scelto')
axes[0].axhline(y=0, color='gray', linewidth=0.5)
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].set_title("sin(θ): monotona su [-π/2, π/2]")

axes[1].plot(theta, np.cos(theta), color='lightgray', label='cos(θ) completo')
theta_decrescente = np.linspace(0, np.pi, 100)
axes[1].plot(theta_decrescente, np.cos(theta_decrescente), color='tab:orange', linewidth=3, label='tratto decrescente scelto')
axes[1].axhline(y=0, color='gray', linewidth=0.5)
axes[1].legend()
axes[1].grid(True, alpha=0.3)
axes[1].set_title("cos(θ): monotona su [0, π]")

plt.tight_layout()
plt.show()

# Verifica che questi intervalli percorrano tutta l'immagine [-1,1]
print("sin(-π/2) =", np.sin(-np.pi/2), "  sin(π/2) =", np.sin(np.pi/2))
print("cos(0) =", np.cos(0), "  cos(π) =", np.cos(np.pi))
````

## Esercizio svolto e commentato

**Problema:** verifica che $\sin x$ è davvero iniettiva sull'intervallo $\left[-\frac\pi2,\frac\pi2\right]$, mostrando che nessun altro punto dello stesso intervallo dà lo stesso valore.

**Ragionamento:**

**Passo 1 — richiamiamo il criterio del notebook 3.9**: se una funzione è strettamente monotona su un intervallo, è automaticamente iniettiva lì. Ci basta quindi confermare che $\sin x$ sia strettamente crescente su $\left[-\frac\pi2,\frac\pi2\right]$ (l'abbiamo enunciato nella teoria, ma vediamo perché è plausibile).

**Passo 2 — pensiamo al punto $P$ sulla circonferenza goniometrica** (notebook 2.9). Per $\theta$ che va da $-\frac\pi2$ a $\frac\pi2$, il punto $P$ percorre la **metà destra** della circonferenza (dove $\cos\theta\ge0$), partendo dal punto più basso $(0,-1)$ e arrivando al punto più alto $(0,1)$, passando per $(1,0)$. Durante questo percorso, l'**ordinata** di $P$ (cioè $\sin\theta$) sale **sempre**, senza mai scendere: è il modo geometrico di vedere la stretta crescenza.

**Passo 3 — confrontiamo con un contro-esempio fuori dall'intervallo scelto.** Se avessimo incluso, per esempio, $\theta=\frac{3\pi}4$ (fuori da $\left[-\frac\pi2,\frac\pi2\right]$): $\sin\left(\frac{3\pi}4\right)=\frac{\sqrt2}2$, lo stesso valore di $\sin\left(\frac\pi4\right)=\frac{\sqrt2}2$ (che invece **è** nell'intervallo scelto). Se avessimo incluso $\frac{3\pi}4$ nel nostro intervallo, avremmo perso l'iniettività — infatti $\frac{3\pi}4$ sta nel tratto in cui il seno **decresce** (dopo aver raggiunto il massimo in $\frac\pi2$), quindi ripete valori già assunti nel tratto crescente.

**Conclusione**: **solo** fermandosi esattamente a $\frac\pi2$ (dove il seno raggiunge il suo massimo, dopo il quale ricomincia a scendere) evitiamo ogni ripetizione di valori: è il confine naturale dell'intervallo di iniettività.

**Perché la scelta dell'intervallo dev'essere così precisa, e non "circa" $\left[-\frac\pi2,\frac\pi2\right]$:** un solo punto in più (come $\frac{3\pi}4$) reintrodurrebbe una ripetizione di valori, rompendo l'iniettività necessaria per costruire l'inversa (notebook 3.6). La scelta dell'intervallo è quindi tassativa, non approssimativa: deve corrispondere esattamente a un tratto di monotonia stretta, né più corto (perderemmo suriettività sull'immagine $[-1,1]$) né più lungo (perderemmo iniettività).

````python
import numpy as np

print("sin(3π/4) =", np.sin(3*np.pi/4))
print("sin(π/4) =", np.sin(np.pi/4))
print("Sono uguali?", np.isclose(np.sin(3*np.pi/4), np.sin(np.pi/4)))
print("\n3π/4 è fuori da [-π/2, π/2]?", 3*np.pi/4 > np.pi/2)
````

## Riepilogo finale

- Nessuna delle tre funzioni trigonometriche è iniettiva sul proprio dominio naturale, a causa della periodicità.
- Intervalli standard di monotonia stretta (che percorrono l'intera immagine): seno su $\left[-\frac\pi2,\frac\pi2\right]$ (crescente); coseno su $[0,\pi]$ (decrescente); tangente su $\left(-\frac\pi2,\frac\pi2\right)$ (crescente).
- La scelta di questi intervalli è tassativa: devono corrispondere esattamente a un tratto di monotonia stretta che copre l'intera immagine, né più né meno.
- Questa restrizione del dominio è il prerequisito indispensabile (nello spirito del notebook 3.6) per costruire le funzioni trigonometriche inverse.
