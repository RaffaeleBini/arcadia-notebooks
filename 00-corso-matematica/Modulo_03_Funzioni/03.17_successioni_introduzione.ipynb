# 3.17 — Successioni (introduzione)

**Prerequisiti:** 3.2 Funzioni reali, 0.1 Numeri naturali.


## Perché ci interessa

Finora abbiamo studiato funzioni $f:D\to\mathbb{R}$ con $D$ un intervallo (o unione di intervalli) di $\mathbb{R}$: un dominio "continuo", senza salti. Ma esiste un caso speciale, altrettanto importante, in cui il dominio è **discreto**: i numeri naturali $\mathbb{N}$ (notebook 0.1). Una funzione $a:\mathbb{N}\to\mathbb{R}$ si chiama **successione**, e descrive un elenco ordinato e infinito di numeri: $a_0,a_1,a_2,\dots$

Le successioni chiudono questo modulo e aprono la porta al prossimo: il Modulo 4 studierà cosa succede a una successione quando l'indice $n$ cresce senza limite — il concetto di **limite**, il cuore di tutta l'analisi matematica.

## Teoria, passo dopo passo

### Definizione

Una **successione** (di numeri reali) è una funzione $a:\mathbb{N}\to\mathbb{R}$ (notebook 3.1). Invece di scrivere $a(n)$ come per una funzione qualsiasi, si usa quasi sempre la notazione con **indice in basso**: $a_n$. L'intera successione si indica $(a_n)_{n\in\mathbb{N}}$, o più semplicemente $(a_n)$.

**Perché il dominio "discreto" cambia le cose**: una funzione reale (notebook 3.2) ha un grafico che è una curva continua (o quasi); il "grafico" di una successione è invece una collezione di **punti isolati** $(0,a_0),(1,a_1),(2,a_2),\dots$, uno per ogni naturale — niente da collegare con una linea, perché tra un naturale e il successivo non ci sono altri punti del dominio.

### Modi per definire una successione

1. **Formula esplicita**: $a_n=\frac1{n+1}$ (usiamo $n+1$ per evitare la divisione per $0$ quando $n=0$). Dato $n$, calcoliamo direttamente $a_n$.
2. **Definizione ricorsiva**: si dà il primo termine (o i primi termini) e una regola per ottenere ogni termine da quelli precedenti. Esempio: $a_0=1$, $a_{n+1}=2a_n$ (ogni termine è il doppio del precedente). Questo produce $1,2,4,8,16,\dots$ — la stessa successione di $a_n=2^n$, ma descritta in modo diverso.

La definizione ricorsiva è concettualmente imparentata con il principio di induzione (notebook 0.20): un "passo base" ($a_0$) e una regola per passare da un termine al successivo, esattamente come il passo induttivo generava $P(k+1)$ da $P(k)$.

### Esempi fondamentali

- **Successione aritmetica**: $a_n=a_0+nd$ (ogni termine si ottiene sommando una quantità fissa $d$, la **ragione**, al precedente). È l'analogo discreto di una retta (notebook 2.2): stesso "incremento" costante ogni passo, come il coefficiente angolare.
- **Successione geometrica**: $a_n=a_0\cdot q^n$ (ogni termine si ottiene moltiplicando per un fattore fisso $q$, la **ragione**). È l'analogo discreto di una funzione esponenziale (notebook 1.11, 3.13): stesso principio di crescita/decrescita moltiplicativa, applicato però solo ai naturali.

### Proprietà ereditate dal notebook 3.2, 3.9, 3.10

Le successioni sono funzioni reali (con dominio $\mathbb{N}$), quindi ereditano senza modifiche i concetti già visti:

- **Monotonia** (notebook 3.9): $(a_n)$ è crescente se $a_n\le a_{n+1}$ per ogni $n$ (l'analogo discreto della definizione generale, con $x_1<x_2$ sostituito da "indici consecutivi").
- **Limitatezza** (notebook 3.10): $(a_n)$ è limitata se esiste $M$ tale che $|a_n|\le M$ per ogni $n$.

Queste due proprietà saranno gli ingredienti chiave, nel Modulo 4, per capire quando una successione "si avvicina" a un valore preciso man mano che $n$ cresce.

````python
import sympy as sp

n = sp.symbols('n', integer=True, nonnegative=True)

# Successione aritmetica: a_n = 3 + 2n (a_0=3, ragione d=2)
a_aritmetica = 3 + 2*n
print("Successione aritmetica (a_0=3, d=2):", [a_aritmetica.subs(n, k) for k in range(6)])

# Successione geometrica: a_n = 5 * (1/2)^n (a_0=5, ragione q=1/2)
a_geometrica = 5 * sp.Rational(1, 2)**n
print("Successione geometrica (a_0=5, q=1/2):", [a_geometrica.subs(n, k) for k in range(6)])

# Definizione ricorsiva: a_0=1, a_(n+1)=2*a_n -> confrontiamo con 2^n
def successione_ricorsiva(N):
    valori = [1]
    for i in range(N):
        valori.append(2 * valori[-1])
    return valori

print("\nSuccessione ricorsiva (a_0=1, a_(n+1)=2a_n):", successione_ricorsiva(5))
print("Confronto con 2^n:", [2**k for k in range(6)])

# Visualizzazione: punti isolati, non una curva continua
import matplotlib.pyplot as plt
indici = list(range(10))
valori = [float(a_geometrica.subs(n, k)) for k in indici]
plt.figure(figsize=(7, 4))
plt.plot(indici, valori, 'o', color='tab:blue')  # solo punti, senza linea che li unisce
plt.title("Successione geometrica: punti isolati, non una curva continua")
plt.xlabel("n"); plt.ylabel("a_n")
plt.grid(True, alpha=0.3)
plt.show()
````

## Esercizio svolto e commentato

**Problema:** la successione definita ricorsivamente da $a_1=2$, $a_{n+1}=a_n+3n$ (per $n\ge1$): calcola i primi termini e trova una formula esplicita.

**Ragionamento:**

**Passo 1 — calcoliamo i primi termini applicando la ricorsione**, un passo alla volta:

$$a_1=2$$
$$a_2=a_1+3(1)=2+3=5$$
$$a_3=a_2+3(2)=5+6=11$$
$$a_4=a_3+3(3)=11+9=20$$

**Passo 2 — cerchiamo uno schema nelle differenze.** Osserviamo che $a_2-a_1=3$, $a_3-a_2=6$, $a_4-a_3=9$: le differenze sono $3,6,9,\dots$, cioè $3(n-1)$ per passare da $a_{n-1}$ ad $a_n$ — coerente con la regola ricorsiva stessa ($a_{n+1}-a_n=3n$).

**Passo 3 — sommiamo tutte le differenze per ricostruire $a_n$ da $a_1$.** Il termine $a_n$ si ottiene da $a_1$ sommando tutte le differenze intermedie:

$$a_n = a_1 + \sum_{k=1}^{n-1} 3k = 2+3\sum_{k=1}^{n-1}k$$

**Passo 4 — usiamo la formula della somma dei primi interi** (dimostrata per induzione nel notebook 0.20!): $\sum_{k=1}^{n-1}k=\frac{(n-1)n}{2}$.

$$a_n = 2+3\cdot\frac{(n-1)n}{2} = 2+\frac{3n^2-3n}{2}$$

**Verifica** su $n=4$: $a_4=2+\frac{3(16)-3(4)}{2}=2+\frac{48-12}{2}=2+18=20$. ✓ Coincide con il calcolo diretto del passo 1.

**Perché "sommare tutte le differenze" è una strategia generale per passare da una definizione ricorsiva a una esplicita, quando la ricorsione è additiva:** ogni passo della ricorsione $a_{n+1}=a_n+(\text{qualcosa})$ aggiunge un pezzo; sommando tutti i pezzi aggiunti dall'inizio, ricostruiamo il valore finale rispetto al primo termine. È lo stesso principio usato nel notebook 0.20 per dimostrare per induzione la formula della somma dei primi $n$ numeri: lì il "qualcosa" era ogni singolo numero naturale.

````python
import sympy as sp

n = sp.symbols('n', integer=True, positive=True)

# Verifica della formula esplicita trovata
formula_esplicita = 2 + sp.Rational(3,2)*n**2 - sp.Rational(3,2)*n
print("Formula esplicita semplificata:", sp.simplify(formula_esplicita))

# Confronto con il calcolo ricorsivo diretto
def ricorsiva(N):
    valori = {1: 2}
    for k in range(1, N):
        valori[k+1] = valori[k] + 3*k
    return valori

valori_ricorsivi = ricorsiva(6)
for k in range(1, 6):
    dalla_formula = formula_esplicita.subs(n, k)
    print(f"n={k}: ricorsivo={valori_ricorsivi[k]}, formula={dalla_formula}")
````

## Riepilogo finale

- Una successione è una funzione $a:\mathbb{N}\to\mathbb{R}$, indicata $(a_n)$: un elenco ordinato e infinito di numeri, con "grafico" fatto di punti isolati, non una curva continua.
- Si definisce con una **formula esplicita** (calcolo diretto di $a_n$) oppure in modo **ricorsivo** (primo termine + regola per ottenere ogni termine dal precedente, nello spirito dell'induzione, notebook 0.20).
- Successione **aritmetica** ($a_n=a_0+nd$): analogo discreto di una retta. Successione **geometrica** ($a_n=a_0q^n$): analogo discreto di un'esponenziale.
- Monotonia e limitatezza (notebook 3.9, 3.10) si applicano senza modifiche alle successioni, e saranno gli strumenti chiave nel Modulo 4 per capire il concetto di limite.
