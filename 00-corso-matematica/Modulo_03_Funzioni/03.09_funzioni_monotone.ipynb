# 3.9 — Funzioni monotone

**Prerequisiti:** 3.5 Funzioni iniettive, suriettive, biiettive, 1.11 Esponenziali.


## Perché ci interessa

Nel notebook 1.11 abbiamo osservato, senza nominarla, una proprietà cruciale delle esponenziali: per $a>1$, $a^x$ "cresce sempre" al crescere di $x$; per $0<a<1$, "decresce sempre". Questa proprietà — la **monotonia** — è così utile che la formalizziamo ora in generale: ci dice come si comporta una funzione "muovendosi" lungo il dominio, ed è strettamente legata all'iniettività (notebook 3.5).

## Teoria, passo dopo passo

### Definizioni

Una funzione $f:D\to\mathbb{R}$ è:

- **Strettamente crescente** se $x_1<x_2 \Rightarrow f(x_1)<f(x_2)$, per ogni $x_1,x_2\in D$.
- **Strettamente decrescente** se $x_1<x_2 \Rightarrow f(x_1)>f(x_2)$.
- **Crescente** (non stretta) se $x_1<x_2 \Rightarrow f(x_1)\le f(x_2)$ (ammette "tratti piatti").
- **Decrescente** (non stretta) se $x_1<x_2 \Rightarrow f(x_1)\ge f(x_2)$.

Una funzione che è (strettamente) crescente o (strettamente) decrescente si dice **monotona** (o "strettamente monotona"). La differenza tra "stretta" e "non stretta" è la stessa distinzione vista per le disuguaglianze nel notebook 1.8 ($<$ contro $\le$): la versione stretta esclude che la funzione resti "ferma" su un tratto.

### Monotonia stretta implica iniettività

Ecco perché la monotonia è così legata al notebook 3.5: se $f$ è strettamente crescente (o decrescente), allora $f$ è **iniettiva**. Dimostrazione diretta: se $x_1\ne x_2$, per la tricotomia dell'ordine (notebook 0.1) deve essere $x_1<x_2$ oppure $x_2<x_1$. In entrambi i casi, la definizione di monotonia stretta garantisce $f(x_1)\ne f(x_2)$ (strettamente minore o maggiore, mai uguale). Quindi input diversi danno sempre output diversi: è la definizione stessa di iniettività.

**Attenzione**: l'implicazione va in un solo verso! Una funzione può essere iniettiva senza essere monotona (esempio nell'esercizio svolto).

### Monotonia locale vs globale

Una funzione può essere monotona su tutto il suo dominio (come $a^x$ con $a>1$, sempre crescente, notebook 1.10-1.11), oppure solo su **alcuni intervalli** (monotonia **locale**). Esempio: $f(x)=x^2$ (notebook 2.5) è strettamente decrescente su $(-\infty,0]$ e strettamente crescente su $[0,+\infty)$, ma **non è monotona** su tutto $\mathbb{R}$ (coerente con il fatto che non è iniettiva su $\mathbb{R}$, notebook 3.5: la non-iniettività globale e la non-monotonia globale vanno di pari passo).

Il punto $x=0$, dove la funzione "cambia direzione" (da decrescente a crescente), è il vertice della parabola (notebook 2.5): non è un caso, è esattamente il punto che separa i due tratti di monotonia.

### Perché la restrizione per l'inversa (notebook 3.6) spesso usa la monotonia

Nel notebook 3.6 abbiamo restretto $x^2$ a $[0,+\infty)$ per renderla biiettiva e trovarne l'inversa. Ora vediamo *perché* funziona: su $[0,+\infty)$, $x^2$ è strettamente crescente, quindi (per quanto appena dimostrato) iniettiva. Restringere il dominio a un intervallo di **monotonia stretta** è la strategia standard per ottenere l'iniettività necessaria a costruire un'inversa — la useremo di nuovo, in modo cruciale, nel notebook 3.16 per le funzioni trigonometriche inverse.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Verifica della monotonia con SymPy: is_increasing / is_decreasing su un intervallo
f = 2**x
print("2^x è crescente su R?", f.is_increasing if hasattr(f, 'is_increasing') else "usa calculus.util")

from sympy.calculus.util import minimum, maximum
from sympy import Interval

# Verifica diretta della definizione su alcuni punti campione
def verifica_monotonia_crescente(funzione, punti):
    ordinati = sorted(punti)
    valori = [funzione.subs(x, p) for p in ordinati]
    return all(valori[i] < valori[i+1] for i in range(len(valori)-1))

campione = [-2, -1, 0, 1, 2, 3]
print("2^x è crescente sul campione?", verifica_monotonia_crescente(2**x, campione))
print("x^2 è crescente sul campione (tutto R)?", verifica_monotonia_crescente(x**2, campione))
print("x^2 è crescente solo su [0,+inf)?", verifica_monotonia_crescente(x**2, [0, 1, 2, 3]))

# Grafico di x^2 con i due tratti di monotonia evidenziati
x_neg = np.linspace(-3, 0, 100)
x_pos = np.linspace(0, 3, 100)
fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(x_neg, x_neg**2, color='tab:red', label='decrescente su (-∞,0]')
ax.plot(x_pos, x_pos**2, color='tab:blue', label='crescente su [0,+∞)')
ax.plot(0, 0, 'ko', label='vertice: punto di svolta')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("x² non è monotona su R, ma lo è su ciascun tratto")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** la funzione $f(x)=\dfrac{1}{x}$ (con dominio $\mathbb{R}\setminus\{0\}$) è iniettiva. È anche monotona su tutto il suo dominio?

**Ragionamento:**

**Passo 1 — verifichiamo l'iniettività** (per completezza, con la definizione per contronominale, notebook 3.5): se $\frac{1}{x_1}=\frac{1}{x_2}$, moltiplicando entrambi i membri per $x_1x_2$ (invariantiva, lecito perché $x_1,x_2\ne0$): $x_2=x_1$. Confermato: $f$ è iniettiva.

**Passo 2 — controlliamo se è monotona su tutto il dominio, con un test diretto.** Proviamo $x_1=-1$ e $x_2=1$ (con $x_1<x_2$): $f(-1)=-1$, $f(1)=1$. Qui $f(x_1)<f(x_2)$, sembra crescente. Proviamo ora $x_1=1$ e $x_2=2$ (ancora $x_1<x_2$): $f(1)=1$, $f(2)=\frac12$. Qui $f(x_1)>f(x_2)$: **decrescente**! Abbiamo trovato un'incoerenza: la funzione non può essere né sempre crescente né sempre decrescente su tutto il dominio.

**Passo 3 — interpretiamo cosa succede.** Il dominio $\mathbb{R}\setminus\{0\}$ non è un intervallo unico: è l'unione di due intervalli separati, $(-\infty,0)$ e $(0,+\infty)$ (notebook 0.13). $f$ è strettamente **decrescente su ciascuno dei due intervalli separatamente** (puoi verificarlo: su $(0,+\infty)$, prendendo $x_1<x_2$ entrambi positivi, $\frac{1}{x_1}>\frac{1}{x_2}$ sempre), ma **non ha senso** confrontare un punto del primo intervallo con uno del secondo nello stesso modo, perché il "salto" attraverso $x=0$ (dove la funzione non è nemmeno definita) rompe la monotonia globale.

**Conclusione**: $f(x)=\frac1x$ è iniettiva su tutto il dominio, ma **non è monotona su tutto il dominio** (lo è solo separatamente su ciascuno dei due intervalli che lo compongono) — l'esempio promesso nella teoria di iniettività senza monotonia globale.

**Perché questo esempio non contraddice l'implicazione "monotona stretta ⟹ iniettiva" vista nella teoria:** l'implicazione va in un solo verso. Non dice che ogni funzione iniettiva sia monotona — dice solo che la monotonia stretta è una condizione **sufficiente** (non necessaria) per l'iniettività. $f(x)=\frac1x$ mostra proprio che esistono funzioni iniettive per altre ragioni (qui, il fatto che il dominio "spezzato" permette due comportamenti monotoni distinti che insieme restano comunque iniettivi).

````python
import sympy as sp

x = sp.symbols('x', real=True, nonzero=True)
f = 1/x

# Test su punti che attraversano lo zero
print("f(-1) =", f.subs(x, -1), " f(1) =", f.subs(x, 1), " f(2) =", f.subs(x, 2))
print("f(-1) < f(1)?", f.subs(x, -1) < f.subs(x, 1))
print("f(1) < f(2)?", f.subs(x, 1) < f.subs(x, 2), " (se fosse crescente dovrebbe essere True)")

# Verifica della monotonia SEPARATA su ciascun intervallo
campione_positivi = [1, 2, 3, 4]
valori_positivi = [f.subs(x, p) for p in campione_positivi]
print("\nSu (0,+inf), valori decrescenti?", all(valori_positivi[i] > valori_positivi[i+1] for i in range(len(valori_positivi)-1)))
````

## Riepilogo finale

- Strettamente crescente: $x_1<x_2 \Rightarrow f(x_1)<f(x_2)$; strettamente decrescente: verso opposto. Le versioni "non strette" ammettono uguaglianza.
- La **monotonia stretta implica l'iniettività** (dimostrabile con la tricotomia dell'ordine), ma non vale il viceversa: esistono funzioni iniettive non monotone su tutto il dominio.
- Una funzione può essere monotona solo **localmente** (su alcuni intervalli), anche se non lo è globalmente: il punto di cambio è spesso un vertice o un punto notevole (come per $x^2$).
- Restringere il dominio a un intervallo di monotonia stretta è la strategia standard per garantire l'iniettività necessaria a costruire una funzione inversa (notebook 3.6).
