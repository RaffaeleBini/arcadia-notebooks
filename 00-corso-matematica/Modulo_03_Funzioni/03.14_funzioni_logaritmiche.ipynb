# 3.14 — Funzioni logaritmiche

**Prerequisiti:** 1.12 Logaritmi, 3.6 Funzione inversa, 3.13 Funzioni esponenziali.


## Perché ci interessa

Nel notebook 3.13 abbiamo stabilito che $f(x)=a^x$, vista come funzione $\mathbb{R}\to(0,+\infty)$, è biiettiva. Per il notebook 3.6, questo garantisce l'esistenza di una funzione inversa: è esattamente $f^{-1}(x)=\log_a x$, il logaritmo introdotto nel notebook 1.12. Ora possiamo dedurre **tutte** le proprietà della funzione logaritmica direttamente dalle proprietà dell'esponenziale, usando la teoria generale delle funzioni inverse — invece di ridimostrarle da zero.

## Teoria, passo dopo passo

### Dominio e immagine, per scambio con l'esponenziale

Dal notebook 3.6: il dominio di $f^{-1}$ è l'immagine di $f$, e viceversa. Applicando questo a $f(x)=a^x$ (dominio $\mathbb{R}$, immagine $(0,+\infty)$, notebook 3.13):

$$\text{Dominio di } \log_a x = (0,+\infty) \qquad \text{Immagine di } \log_a x = \mathbb{R}$$

Coerente con quanto già sapevamo dal notebook 1.12 (il logaritmo richiede argomento positivo), ma ora otteniamo anche l'immagine "gratis": il logaritmo è **suriettivo** su $\mathbb{R}$ (assume ogni valore reale), a differenza dell'esponenziale che non lo era.

### Grafico: simmetria rispetto a $y=x$

Dal notebook 3.6: il grafico di $\log_a x$ è il simmetrico del grafico di $a^x$ rispetto alla retta $y=x$. Questo spiega visivamente perché il logaritmo "cresce lentissimamente": l'esponenziale cresce molto rapidamente lungo l'asse $y$, quindi la sua riflessione cresce molto lentamente lungo l'asse $x$.

### Monotonia (ereditata dall'esponenziale)

Un fatto generale sulle funzioni inverse: se $f$ è strettamente crescente, anche $f^{-1}$ lo è (e analogamente per decrescente) — si può dimostrare per assurdo (notebook 0.20): se $f^{-1}$ non fosse crescente, esisterebbero $y_1<y_2$ con $f^{-1}(y_1)\ge f^{-1}(y_2)$; applicando $f$ (crescente) a entrambi i membri, otterremmo $y_1\ge y_2$, contraddicendo l'ipotesi $y_1<y_2$. Quindi:

$$\log_a x \text{ è strettamente crescente se } a>1, \text{ strettamente decrescente se } 0<a<1$$

(stesso segno di monotonia dell'esponenziale corrispondente).

### Asintoto verticale

L'esponenziale ha un asintoto **orizzontale** in $y=0$ (notebook 3.13): per simmetria rispetto a $y=x$, il logaritmo ha un asintoto **verticale** in $x=0$ (l'asse $y$ diventa l'asse $x$, scambiando i ruoli). Avvicinandosi a $x=0^+$ (da destra, unico verso possibile dato il dominio), $\log_a x\to-\infty$ se $a>1$.

### Punti notevoli, per simmetria

L'esponenziale passa sempre per $(0,1)$ (notebook 1.11); per simmetria rispetto a $y=x$, il logaritmo passa sempre per $(1,0)$ — coerente con $\log_a 1=0$ per ogni base (notebook 1.12).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', positive=True)

# Verifica dominio/immagine tramite scambio con l'esponenziale
from sympy.calculus.util import function_range
immagine_log = function_range(sp.log(x, 2), x, sp.Interval.open(0, sp.oo))
print("Immagine di log_2(x):", immagine_log)

# Grafico: esponenziale, logaritmo, e la simmetria rispetto a y=x
x_vals_esp = np.linspace(-4, 4, 200)
x_vals_log = np.linspace(0.05, 16, 200)

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x_vals_esp, 2**x_vals_esp, label='2^x')
ax.plot(x_vals_log, np.log2(x_vals_log), label='log_2(x)')
ax.plot(np.linspace(-4, 16, 10), np.linspace(-4, 16, 10), '--', color='gray', label='y=x')
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.plot(0, 1, 'ko'); ax.plot(1, 0, 'ko')
ax.legend()
ax.set_aspect('equal')
ax.set_xlim(-4, 8); ax.set_ylim(-4, 8)
ax.grid(True, alpha=0.3)
ax.set_title("2^x e log_2(x): simmetrici rispetto a y=x")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** confronta la crescita di $f(x)=\log_2 x$ e $g(x)=\sqrt{x}$ per $x$ grande: quale cresce più lentamente?

**Ragionamento:**

**Passo 1 — confrontiamo alcuni valori:**

| $x$ | $\log_2 x$ | $\sqrt{x}$ |
|---|---|---|
| $100$ | $\approx6.64$ | $10$ |
| $10000$ | $\approx13.29$ | $100$ |
| $1000000$ | $\approx19.93$ | $1000$ |

**Passo 2 — osserviamo il divario.** Man mano che $x$ cresce (anche di molto, moltiplicandolo per $100$ ogni riga), $\sqrt{x}$ cresce proporzionalmente (di un fattore $10$ ogni volta, essendo $\sqrt{100x}=10\sqrt x$), mentre $\log_2 x$ cresce solo **additivamente** di poco (perché $\log_2(100x)=\log_2 100+\log_2 x$, proprietà del prodotto, notebook 1.12: moltiplicare l'argomento per $100$ si traduce in una **somma** costante sul logaritmo, non in una moltiplicazione).

**Conclusione**: il logaritmo cresce, per $x$ grande, più lentamente di **qualunque** potenza $x^k$ (anche per $k$ piccolissimo, come $k=\frac12$ qui). È l'esatto **opposto** del comportamento dell'esponenziale visto nel notebook 3.13 (che batteva ogni potenza): il logaritmo è, in un certo senso, "agli antipodi" delle potenze in termini di velocità di crescita — coerente con il fatto che è l'inverso dell'esponenziale, e l'esponenziale è "agli antipodi opposti".

**Perché il comportamento moltiplicativo dell'argomento diventa additivo sul logaritmo è la chiave di questa lentezza:** ogni volta che l'input viene moltiplicato per un fattore fisso (anche enorme, come $100$ o $1000000$), il logaritmo aumenta solo di una quantità **fissa** (il logaritmo di quel fattore). Questo è precisamente il motivo per cui il logaritmo "compressa" numeri che crescono esponenzialmente in una scala che cresce linearmente — la stessa idea usata praticamente nella scala Richter o nei decibel.

````python
import math

for x_val in [100, 10000, 1000000]:
    log_val = math.log2(x_val)
    radice_val = math.sqrt(x_val)
    print(f"x={x_val}: log_2(x)={log_val:.2f}, sqrt(x)={radice_val:.2f}")
````
