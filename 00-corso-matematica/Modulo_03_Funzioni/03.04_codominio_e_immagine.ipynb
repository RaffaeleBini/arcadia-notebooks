# 3.4 — Codominio e immagine

**Prerequisiti:** 3.3 Dominio, 2.5 Parabola.


## Perché ci interessa

Consideriamo $f:\mathbb{R}\to\mathbb{R}$, $f(x)=x^2$. Il codominio che abbiamo dichiarato è $\mathbb{R}$ — ma la funzione **assume davvero** ogni valore reale? No: $x^2$ non è mai negativo. Il codominio è l'insieme "dichiarato" di arrivo, ma i valori **effettivamente raggiunti** sono un sottoinsieme, spesso più piccolo: l'**immagine**. Distinguere questi due concetti, spesso confusi, è essenziale per capire bene cosa significa che una funzione è "suriettiva" (notebook 3.5).

## Teoria, passo dopo passo

### Definizione di immagine

Data $f:D\to B$, l'**immagine** di $f$ (indicata $\text{Im}(f)$ o $f(D)$) è l'insieme di tutti i valori effettivamente assunti:

$$\text{Im}(f) = \{f(x) : x\in D\} = \{y\in B : \exists x\in D,\ f(x)=y\}$$

(la seconda scrittura usa il quantificatore esistenziale, notebook 0.16: $y$ è nell'immagine se esiste **almeno un** $x$ che ci arriva). Sempre vale $\text{Im}(f)\subseteq B$ (notebook 0.12): l'immagine è sempre un sottoinsieme del codominio dichiarato, non può mai "uscirne" (per definizione di funzione, ogni $f(x)$ appartiene a $B$).

### Codominio "dichiarato" vs "effettivo"

Quando scriviamo $f:\mathbb{R}\to\mathbb{R}$, il codominio $\mathbb{R}$ è spesso una scelta **conveniente e generosa** (per non doverci preoccupare, mentre definiamo la funzione, di quali valori userà davvero), non necessariamente il più "stretto" possibile. L'immagine, invece, è **determinata dalla funzione stessa**: non è una scelta, va calcolata.

Esempio: per $f(x)=x^2$ (dominio $\mathbb{R}$, codominio dichiarato $\mathbb{R}$), l'immagine è $\text{Im}(f)=[0,+\infty)$: ogni numero non negativo è raggiunto (per $y\ge0$, basta prendere $x=\sqrt y$), ma nessun numero negativo lo è mai.

### Trovare l'immagine: la domanda da porsi

Per trovare l'immagine di $f$, ci chiediamo: **per quali $y$ l'equazione $f(x)=y$ ha almeno una soluzione $x\in D$?** Questo trasforma il problema "trovare l'immagine" in un problema già familiare: risolvere un'equazione (nell'incognita $x$), trattando $y$ come un parametro (proprio come nel notebook 1.5, esercizio 4, dove risolvevamo un'equazione al variare di un parametro $k$).

Esempio, ripreso da $f(x)=x^2$: risolviamo $x^2=y$ per $x$. Se $y\ge0$: $x=\pm\sqrt y$, soluzioni esistenti. Se $y<0$: nessuna soluzione reale (notebook 0.4, la radice di un negativo non è reale). Quindi $\text{Im}(f)=[0,+\infty)$, confermando quanto detto sopra.

### Immagine tramite il grafico

Visivamente, l'immagine di $f$ corrisponde a tutti i valori $y$ tali che la retta orizzontale $y=k$ interseca il grafico **almeno una volta** — un test speculare a quello della retta verticale (notebook 0.19), ma applicato all'asse delle $y$ invece che delle $x$. Per la parabola $y=x^2$ (notebook 2.5), le rette orizzontali $y=k$ con $k\ge0$ intersecano sempre il grafico (in uno o due punti, a seconda che $k=0$ o $k>0$), mentre quelle con $k<0$ non lo intersecano mai: coerente con $\text{Im}(f)=[0,+\infty)$.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y', real=True)

# Troviamo l'immagine di f(x)=x^2 risolvendo x^2=y per x
soluzioni = sp.solve(sp.Eq(x**2, y), x)
print("Soluzioni di x^2=y in funzione di y:", soluzioni)
print("Esistono soluzioni reali solo se y >= 0")

# Verifica con SymPy: calcolo diretto dell'immagine
from sympy.calculus.util import function_range
immagine = function_range(x**2, x, sp.S.Reals)
print("\nImmagine calcolata da SymPy:", immagine)

# Visualizzazione: il test della retta orizzontale
x_vals = np.linspace(-3, 3, 200)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x_vals, x_vals**2, label='f(x)=x²')
ax.axhline(y=4, color='green', linestyle='--', label='y=4 (interseca: 4 è nell\'immagine)')
ax.axhline(y=-2, color='red', linestyle='--', label='y=-2 (non interseca: -2 non è nell\'immagine)')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_ylim(-4, 8)
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova l'immagine di $f(x)=\dfrac{1}{x^2+1}$ (dominio $\mathbb{R}$, poiché $x^2+1$ non si annulla mai).

**Ragionamento:**

**Passo 1 — impostiamo l'equazione** $f(x)=y$, cioè $\dfrac{1}{x^2+1}=y$, e ci chiediamo per quali $y$ esiste una soluzione $x$ reale.

**Passo 2 — isoliamo $x^2$.** Prima notiamo che $y\ne0$ (la frazione non può mai valere $0$: un numeratore $1$ non si annulla mai). Per $y\ne0$, invertiamo (notebook 0.3):

$$x^2+1=\frac{1}{y} \quad\Rightarrow\quad x^2=\frac{1}{y}-1$$

**Passo 3 — troviamo per quali $y$ questa equazione ha soluzione reale.** Serve $x^2\ge0$ (un quadrato non è mai negativo, notebook 0.5), quindi:

$$\frac{1}{y}-1\ge0 \quad\Rightarrow\quad \frac{1}{y}\ge1$$

**Passo 4 — risolviamo questa disequazione fratta** (notebook 1.9). Riscriviamo come $\frac1y-1\ge0$, cioè $\frac{1-y}{y}\ge0$. Studiamo il segno: numeratore $1-y\ge0 \Leftrightarrow y\le1$; denominatore $y>0$. Schema dei segni: positivo per $0<y\le1$ (numeratore e denominatore concordi), negativo altrove (per $y<0$ o $y>1$), con $y=0$ escluso (denominatore).

**Conclusione**: l'immagine è $(0,1]$.

**Interpretazione**: il valore massimo $y=1$ si raggiunge per $x=0$ ($f(0)=1$), e $f(x)$ si avvicina a $0$ senza mai raggiungerlo quando $|x|$ cresce (un comportamento asintotico, che approfondiremo con i limiti nel Modulo 4).

**Perché la condizione "$x^2\ge0$" (non $x^2>0$) è quella corretta da imporre, e non un'altra:** stiamo chiedendo se esiste **un** valore di $x$ (reale, qualsiasi, incluso $0$) che soddisfi l'equazione: $x^2=0$ è una soluzione perfettamente valida (corrisponde a $x=0$). La condizione di esistenza di una soluzione reale per $x^2=k$ è sempre $k\ge0$, non $k>0$ — un dettaglio che va sempre controllato con attenzione.

````python
import sympy as sp
from sympy.calculus.util import function_range

x = sp.symbols('x', real=True)
f = 1 / (x**2 + 1)
immagine = function_range(f, x, sp.S.Reals)
print("Immagine:", immagine)

# Verifica: valore massimo in x=0
print("f(0) =", f.subs(x, 0))
````

## Riepilogo finale

- L'**immagine** $\text{Im}(f)$ è l'insieme dei valori effettivamente assunti da $f$; è sempre un sottoinsieme del codominio dichiarato, spesso più piccolo.
- Per trovare l'immagine, si risolve $f(x)=y$ nell'incognita $x$, trattando $y$ come parametro, e si determina per quali $y$ esiste almeno una soluzione.
- Visivamente, l'immagine corrisponde ai valori $y$ per cui la retta orizzontale $y=k$ interseca il grafico almeno una volta.
- Il codominio è una scelta (spesso $\mathbb{R}$, per comodità); l'immagine è una proprietà intrinseca della funzione, da calcolare.
