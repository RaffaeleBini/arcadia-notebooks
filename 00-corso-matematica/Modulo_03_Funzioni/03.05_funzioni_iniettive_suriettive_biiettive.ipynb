# 3.5 — Funzioni iniettive, suriettive, biiettive

**Prerequisiti:** 3.4 Codominio e immagine, 0.16 Quantificatori ∀ e ∃.


## Perché ci interessa

Nel notebook 3.4 abbiamo distinto codominio (dichiarato) e immagine (effettiva). Questa distinzione porta naturalmente a due domande sul comportamento di una funzione: "può succedere che due input diversi diano lo stesso output?" e "ogni elemento del codominio viene raggiunto?". Le risposte classificano le funzioni in categorie — iniettive, suriettive, biiettive — che determinano, tra le altre cose, se una funzione può essere "invertita" (il tema del prossimo notebook).

## Teoria, passo dopo passo

### Funzioni iniettive

Una funzione $f:D\to B$ è **iniettiva** se elementi diversi del dominio hanno sempre immagini diverse:

$$\forall x_1,x_2\in D,\quad x_1\ne x_2 \ \Rightarrow\ f(x_1)\ne f(x_2)$$

Equivalentemente (per contronominale, notebook 0.15): $f(x_1)=f(x_2) \Rightarrow x_1=x_2$. Questa seconda forma è spesso più comoda da usare nelle dimostrazioni, perché parte da un'uguaglianza (più facile da manipolare algebricamente) invece che da una disuguaglianza.

**Test grafico**: una funzione è iniettiva se e solo se ogni retta **orizzontale** interseca il grafico **al massimo una volta** (mai due o più) — il complemento del test usato nel notebook 3.4 per l'immagine.

Esempio: $f(x)=2x+1$ (una retta, notebook 2.2) è iniettiva: se $2x_1+1=2x_2+1$, sottraendo $1$ e dividendo per $2$ (invariantiva, notebook 0.5), $x_1=x_2$. Esempio di funzione **non** iniettiva: $f(x)=x^2$, perché $f(-2)=f(2)=4$ pur essendo $-2\ne2$.

### Funzioni suriettive

Una funzione $f:D\to B$ è **suriettiva** se **ogni** elemento del codominio è raggiunto da almeno un elemento del dominio:

$$\forall y\in B,\ \exists x\in D,\quad f(x)=y$$

In altre parole (collegandoci al notebook 3.4): $f$ è suriettiva se e solo se $\text{Im}(f)=B$ (l'immagine coincide con **tutto** il codominio dichiarato, non solo una sua parte).

**Attenzione**: la suriettività dipende dalla scelta del codominio $B$! $f(x)=x^2$, vista come $f:\mathbb{R}\to\mathbb{R}$, **non** è suriettiva (l'immagine è $[0,+\infty)$, non tutto $\mathbb{R}$, notebook 3.4). Ma la stessa formula, vista come $f:\mathbb{R}\to[0,+\infty)$ (restringendo il codominio all'immagine), **è** suriettiva. Questo mostra che iniettività e suriettività non sono proprietà di una "formula" isolata, ma di una funzione **con il suo dominio e codominio specificati**.

### Funzioni biiettive

Una funzione **iniettiva e suriettiva contemporaneamente** si dice **biiettiva** (o biunivoca). Ogni elemento del codominio è raggiunto da **esattamente un** elemento del dominio: né zero (suriettività), né più di uno (iniettività). Le funzioni biiettive sono precisamente quelle che ammettono una funzione inversa (notebook 3.6): la corrispondenza $x\leftrightarrow f(x)$ è "perfetta" in entrambe le direzioni.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Verifica dell'iniettività per contronominale: f(x1)=f(x2) => x1=x2 ?
x1, x2 = sp.symbols('x1 x2', real=True)

f = 2*x + 1
equazione = sp.Eq(f.subs(x, x1), f.subs(x, x2))
soluzione = sp.solve(equazione, x1)
print("Se f(x1)=f(x2), allora x1 =", soluzione, " (coincide con x2? è iniettiva)")

g = x**2
equazione_g = sp.Eq(g.subs(x, x1), g.subs(x, x2))
soluzione_g = sp.solve(equazione_g, x1)
print("\nPer g(x)=x^2, se g(x1)=g(x2), allora x1 =", soluzione_g, " (due possibilità: NON iniettiva)")

# Test grafico: retta orizzontale
fig, axes = plt.subplots(1, 2, figsize=(11, 5))
x_vals = np.linspace(-3, 3, 200)

axes[0].plot(x_vals, 2*x_vals + 1)
axes[0].axhline(y=3, color='red', linestyle='--')
axes[0].set_title("f(x)=2x+1: iniettiva\n(ogni orizzontale tocca 1 punto)")
axes[0].grid(True, alpha=0.3)

axes[1].plot(x_vals, x_vals**2)
axes[1].axhline(y=4, color='red', linestyle='--')
axes[1].plot([-2, 2], [4, 4], 'ro')
axes[1].set_title("f(x)=x²: NON iniettiva\n(la stessa orizzontale tocca 2 punti)")
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $f:\mathbb{R}\to\mathbb{R}$, $f(x)=x^3-x$ è iniettiva.

**Ragionamento:**

**Passo 1 — cerchiamo di applicare la definizione per contronominale**: supponiamo $f(x_1)=f(x_2)$ e vediamo se siamo costretti a concludere $x_1=x_2$.

$$x_1^3-x_1 = x_2^3-x_2$$

**Passo 2 — riorganizziamo**, portando tutto da un lato (invariantiva, notebook 0.5):

$$x_1^3-x_2^3 = x_1-x_2$$

**Passo 3 — scomponiamo il lato sinistro.** $x_1^3-x_2^3$ è una differenza di cubi, che si scompone come $(x_1-x_2)(x_1^2+x_1x_2+x_2^2)$ (un prodotto notevole non ancora visto esplicitamente nel notebook 1.2, ma verificabile sviluppando con la distributiva, notebook 1.1):

$$(x_1-x_2)(x_1^2+x_1x_2+x_2^2) = x_1-x_2$$

**Passo 4 — invece di dividere subito per $(x_1-x_2)$ (rischioso: potrebbe essere zero), raccogliamo:**

$$(x_1-x_2)(x_1^2+x_1x_2+x_2^2) - (x_1-x_2) = 0 \quad\Rightarrow\quad (x_1-x_2)\big(x_1^2+x_1x_2+x_2^2-1\big)=0$$

**Passo 5 — applichiamo la legge di annullamento del prodotto** (notebook 1.6): $x_1-x_2=0$ (cioè $x_1=x_2$, quello che vorremmo) **oppure** $x_1^2+x_1x_2+x_2^2-1=0$. Questa seconda possibilità **non è impossibile**: per esempio $x_1=1,x_2=0$ la soddisfa ($1+0+0-1=0$), pur essendo $x_1\ne x_2$!

**Verifica diretta**: $f(1)=1^3-1=0$ e $f(0)=0^3-0=0$: infatti $f(1)=f(0)=0$ con $1\ne0$.

**Conclusione**: $f(x)=x^3-x$ **non è iniettiva** (abbiamo trovato un controesempio esplicito, nello spirito del notebook 0.15-0.16: basta un solo controesempio per confutare l'iniettività).

**Perché non potevamo "semplicemente dividere per $(x_1-x_2)$" al passo 3:** dividere per un'espressione che potrebbe essere zero è un errore algebrico pericoloso (notebook 0.3): avremmo perso proprio il caso $x_1\ne x_2$ che stavamo cercando di escludere o confermare. Raccogliere e applicare la legge di annullamento del prodotto, invece, tiene conto di **entrambe** le possibilità senza escluderne nessuna a priori.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = x**3 - x

print("f(1) =", f.subs(x, 1))
print("f(0) =", f.subs(x, 0))
print("f(1) == f(0) con 1 != 0?  Quindi NON iniettiva.")
````

## Riepilogo finale

- **Iniettiva**: input diversi danno sempre output diversi ($f(x_1)=f(x_2)\Rightarrow x_1=x_2$); test grafico: ogni retta orizzontale tocca al massimo un punto.
- **Suriettiva**: ogni elemento del codominio è raggiunto, cioè $\text{Im}(f)=B$; dipende dalla scelta del codominio dichiarato, non solo dalla formula.
- **Biiettiva**: iniettiva e suriettiva insieme — corrispondenza perfetta tra dominio e codominio.
- Per dimostrare la non iniettività (o non suriettività) basta un controesempio; per dimostrare l'iniettività serve un argomento generale (tipicamente la definizione per contronominale).
