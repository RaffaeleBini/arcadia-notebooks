# 3.6 — Funzione inversa

**Prerequisiti:** 3.5 Funzioni iniettive, suriettive, biiettive, 1.12 Logaritmi.


## Perché ci interessa

Nel notebook 1.12 abbiamo introdotto il logaritmo come l'operazione che "disfa" l'esponenziale: se $y=a^x$, allora $x=\log_a y$. Questo è il primo esempio (non nominato esplicitamente allora) di **funzione inversa**: una funzione che "torna indietro" da $y$ a $x$. Ora formalizziamo questa idea in generale, usando la biiettività (notebook 3.5) come condizione che la rende possibile.

## Teoria, passo dopo passo

### Definizione

Sia $f:D\to B$ una funzione **biiettiva** (notebook 3.5). La **funzione inversa** $f^{-1}:B\to D$ è definita da:

$$f^{-1}(y)=x \quad\Longleftrightarrow\quad f(x)=y$$

Perché questa definizione ha senso solo se $f$ è biiettiva? Perché dobbiamo poter assegnare a **ogni** $y\in B$ un **unico** $x$: la suriettività garantisce che un tale $x$ esista sempre (nessun $y$ resta "senza risposta"), l'iniettività garantisce che sia **unico** (nessuna ambiguità) — esattamente i due requisiti della definizione di funzione stessa (notebook 0.19), applicati ora a $f^{-1}$.

**Attenzione alla notazione**: $f^{-1}$ **non** significa $\frac{1}{f}$ (il reciproco)! È solo un simbolo per "la funzione inversa", storicamente sfortunato per la somiglianza con l'esponente $-1$, ma concettualmente un'operazione completamente diversa dal reciproco.

### Le proprietà fondamentali

Dalla definizione seguono due identità, che valgono per ogni $x\in D$ e $y\in B$:

$$f^{-1}(f(x)) = x \qquad f(f^{-1}(y)) = y$$

("applicare $f$ e poi $f^{-1}$", o viceversa, riporta sempre al punto di partenza — la useremo nel notebook 3.7 sulla composizione). Per esponenziale e logaritmo (notebook 1.12), questo dà esattamente $\log_a(a^x)=x$ e $a^{\log_a y}=y$: le identità viste allora erano già un caso particolare di questa proprietà generale.

### Il grafico dell'inversa: simmetria rispetto a $y=x$

Se $(x_0,y_0)$ è un punto del grafico di $f$ (cioè $f(x_0)=y_0$), allora $(y_0,x_0)$ è un punto del grafico di $f^{-1}$ (perché $f^{-1}(y_0)=x_0$, per definizione). Il punto $(y_0,x_0)$ si ottiene da $(x_0,y_0)$ **scambiando le coordinate**, un'operazione che corrisponde geometricamente a **riflettere rispetto alla retta $y=x$** (la bisettrice del I e III quadrante, notebook 2.1). Quindi:

> Il grafico di $f^{-1}$ è il simmetrico del grafico di $f$ rispetto alla retta $y=x$.

### Come si trova (algebricamente) l'inversa

Dato $y=f(x)$, per trovare $f^{-1}$ si **risolve rispetto a $x$**, poi si scambiano i nomi delle variabili (per convenzione, la variabile indipendente della funzione inversa si chiama di nuovo $x$):

1. Scrivi $y=f(x)$.
2. Risolvi per $x$ in funzione di $y$: ottieni $x=g(y)$.
3. Scambia i nomi: $f^{-1}(x)=g(x)$.

Esempio: $f(x)=2x+3$. Passo 1: $y=2x+3$. Passo 2: $x=\frac{y-3}{2}$ (invariantiva, notebook 0.5, esattamente come risolvere un'equazione di primo grado, notebook 1.5). Passo 3: $f^{-1}(x)=\frac{x-3}{2}$.

### Restringere il dominio per ottenere la biiettività

Cosa succede se $f$ non è biiettiva, come $f(x)=x^2$ su $\mathbb{R}$ (notebook 3.5)? Una strategia comune è **restringere il dominio** a una parte dove $f$ diventa iniettiva. Per $f(x)=x^2$ ristretta a $[0,+\infty)$: qui è biiettiva (verso $[0,+\infty)$, notebook 3.5, esercizio 3), e la sua inversa è $f^{-1}(x)=\sqrt x$ — non a caso, la radice quadrata è definita apposta per restituire il valore **non negativo** (notebook 0.6): è l'inversa della restrizione di $x^2$ alla semiretta positiva. Questa strategia tornerà cruciale nel notebook 3.16, per definire le funzioni trigonometriche inverse.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y = sp.symbols('x y', real=True)

# Troviamo l'inversa di f(x)=2x+3 algebricamente
f = 2*x + 3
x_in_funzione_di_y = sp.solve(sp.Eq(y, f), x)[0]
print("f(x) =", f)
print("Risolvendo y=f(x) per x:", x_in_funzione_di_y)
f_inversa = x_in_funzione_di_y.subs(y, x)
print("f^-1(x) =", f_inversa)

# Verifica delle identità fondamentali
print("\nVerifica f^-1(f(x)) = x:", sp.simplify(f_inversa.subs(x, f) - x))
print("Verifica f(f^-1(x)) = x:", sp.simplify(f.subs(x, f_inversa) - x))

# Grafico: f, la sua inversa, e la simmetria rispetto a y=x
x_vals = np.linspace(-5, 5, 100)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x_vals, 2*x_vals + 3, label='f(x)=2x+3')
ax.plot(x_vals, (x_vals - 3)/2, label='f⁻¹(x)=(x-3)/2')
ax.plot(x_vals, x_vals, '--', color='gray', label='y=x')
ax.legend()
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_xlim(-5, 5); ax.set_ylim(-5, 5)
ax.set_title("f e f⁻¹ sono simmetriche rispetto a y=x")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** trova l'inversa di $f(x)=\dfrac{2x-1}{x+3}$ (con dominio opportuno, $x\ne-3$), e verifica il risultato.

**Ragionamento:**

**Passo 1 — scriviamo $y=f(x)$:**

$$y=\frac{2x-1}{x+3}$$

**Passo 2 — risolviamo per $x$.** Moltiplichiamo entrambi i membri per $(x+3)$ (invariantiva, notebook 0.5, lecito perché $x\ne-3$):

$$y(x+3) = 2x-1 \quad\Rightarrow\quad yx+3y=2x-1$$

Raggruppiamo i termini con $x$ da un lato (notebook 1.5):

$$yx-2x = -1-3y \quad\Rightarrow\quad x(y-2)=-1-3y$$

Isoliamo $x$ (dividendo per $y-2$, richiede $y\ne2$):

$$x=\frac{-1-3y}{y-2} = \frac{3y+1}{2-y}$$

(abbiamo moltiplicato numeratore e denominatore per $-1$ per una forma più leggibile, notebook 0.3).

**Passo 3 — scambiamo i nomi delle variabili:**

$$f^{-1}(x) = \frac{3x+1}{2-x}$$

**Verifica**, componendo $f$ e $f^{-1}$ (dovrebbe restituire $x$, teoria):

$$f(f^{-1}(x)) = \frac{2\cdot\frac{3x+1}{2-x}-1}{\frac{3x+1}{2-x}+3}$$

Moltiplicando numeratore e denominatore per $(2-x)$ per eliminare le frazioni interne (notebook 1.5): numeratore $2(3x+1)-(2-x)=6x+2-2+x=7x$; denominatore $(3x+1)+3(2-x)=3x+1+6-3x=7$. Risultato: $\frac{7x}{7}=x$. ✓

**Perché il dominio di $f^{-1}$ risulta escludere $x=2$ (osservato al passo 2, $y\ne2$):** $y=2$ corrisponderebbe a un valore che $f$ **non raggiunge mai** (l'immagine di $f$ esclude $2$): infatti se provassimo a risolvere $\frac{2x-1}{x+3}=2$, otterremmo $2x-1=2x+6$, cioè $-1=6$, impossibile (notebook 1.5). Il dominio della funzione inversa è sempre l'**immagine** della funzione originale (notebook 3.4): coerente con lo scambio di ruoli tra dominio e codominio.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = (2*x - 1) / (x + 3)
f_inversa = sp.Rational(3, 1)*x/1  # placeholder, calcoliamo con solve
y = sp.symbols('y', real=True)
x_risolto = sp.solve(sp.Eq(y, f), x)
print("x in funzione di y:", x_risolto)
f_inversa = x_risolto[0].subs(y, x)
print("f^-1(x) =", sp.simplify(f_inversa))

# Verifica della composizione
composizione = sp.simplify(f.subs(x, f_inversa))
print("f(f^-1(x)) =", composizione)
````

**1.** $y=5x-2 \Rightarrow x=\dfrac{y+2}{5}$. Quindi $f^{-1}(x)=\dfrac{x+2}{5}$.

## Riepilogo finale

- Solo le funzioni **biiettive** (notebook 3.5) ammettono inversa: la suriettività garantisce l'esistenza di $f^{-1}(y)$ per ogni $y$, l'iniettività ne garantisce l'unicità.
- $f^{-1}(f(x))=x$ e $f(f^{-1}(x))=x$: applicare la funzione e la sua inversa, in qualunque ordine, riporta al punto di partenza (già visto per esponenziale/logaritmo, notebook 1.12).
- Il grafico di $f^{-1}$ è il simmetrico del grafico di $f$ rispetto alla retta $y=x$.
- Per trovare $f^{-1}$ algebricamente: scrivi $y=f(x)$, risolvi per $x$, scambia i nomi delle variabili. Se $f$ non è biiettiva, si può restringere il dominio per renderla tale.
