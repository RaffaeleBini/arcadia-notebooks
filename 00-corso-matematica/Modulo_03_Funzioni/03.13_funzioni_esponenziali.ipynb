## Teoria, passo dopo passo

### Classificazione completa

Riprendiamo $f(x)=a^x$ ($a>0$, $a\ne1$) e applichiamo tutti gli strumenti del modulo:

- **Dominio** (notebook 3.3): $\mathbb{R}$ (notebook 1.10, estensione a esponenti reali).
- **Immagine** (notebook 3.4): $(0,+\infty)$ — mai zero, mai negativa.
- **Iniettiva** (notebook 3.5): sì, sempre (conseguenza della monotonia stretta, notebook 3.9).
- **Suriettiva su $\mathbb{R}\to\mathbb{R}$?** No: l'immagine $(0,+\infty)$ non è tutto $\mathbb{R}$. Ma è suriettiva se il codominio viene ristretto a $(0,+\infty)$ (coerente col notebook 3.5).
- **Biiettiva** su $\mathbb{R}\to(0,+\infty)$: sì — questo è ciò che permette di costruire l'inversa (il logaritmo, notebook 1.12), come vedremo formalmente nel prossimo notebook.
- **Monotona** (notebook 3.9): strettamente crescente se $a>1$, strettamente decrescente se $0<a<1$.
- **Pari o dispari** (notebook 3.8): né l'una né l'altra in generale (verificalo: $a^{-x}=\frac{1}{a^x}$, che non è uguale né a $a^x$ né a $-a^x$, tranne casi degeneri).
- **Limitata** (notebook 3.10): limitata **inferiormente** (da $0$, mai raggiunto), **non** limitata superiormente (se $a>1$; simmetricamente per $0<a<1$, guardando $x\to-\infty$).

### Il ruolo speciale della base $e$

Tra tutte le basi, $a=e\approx2.71828$ (notebook 1.11) gode di una proprietà che la rende privilegiata in analisi: la sua **rapidità di crescita** in un punto è proporzionale al valore stesso della funzione, con costante di proporzionalità esattamente $1$ — un fatto che avrà senso preciso solo dopo aver definito la derivata (Modulo 5), ma che possiamo già osservare numericamente: il rapporto incrementale di $e^x$ vicino a un punto è molto vicino al valore di $e^x$ in quel punto, più precisamente di quanto accada per qualunque altra base.

### Famiglie di funzioni esponenziali trasformate

Applicando le trasformazioni del grafico viste nel notebook 3.2 a $f(x)=a^x$, otteniamo famiglie di funzioni che modellano fenomeni reali:

- $f(x)=k\cdot a^x$ (con $k>0$): scala verticalmente, cambia il valore in $x=0$ (che diventa $k$, invece di $1$), ma non cambia monotonia né asintoto.
- $f(x)=a^x+c$: trasla verticalmente, cambiando l'asintoto orizzontale da $y=0$ a $y=c$.
- $f(x)=a^{-x}=\left(\frac1a\right)^x$: ribalta rispetto all'asse $y$ (notebook 3.2), scambiando crescita e decrescita.

Queste trasformazioni sono esattamente lo strumento che useremo per modellare crescita/decadimento con condizioni iniziali diverse da $1$ (come nell'esempio dei batteri o del capitale, notebook 1.11).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Classificazione automatica con SymPy
from sympy.calculus.util import function_range
f = 2**x
print("Dominio: R (nessuna restrizione)")
print("Immagine:", function_range(f, x, sp.S.Reals))

# Confronto della "rapidità di crescita relativa" per basi diverse
# (rapporto incrementale approssimato vicino a x=1)
h = 0.0001
for base in [2, sp.E, 3]:
    fb = base**x
    rapporto = (fb.subs(x, 1+h) - fb.subs(x, 1)) / h
    valore_in_1 = fb.subs(x, 1)
    print(f"\nbase={base}: f(1)={float(valore_in_1):.4f}, rapporto incrementale in x=1 ≈ {float(rapporto):.4f}")
    print(f"  rapporto/f(1) = {float(rapporto/valore_in_1):.4f}  (per base=e, dovrebbe essere ≈1)")

# Grafico delle trasformazioni
x_vals = np.linspace(-3, 3, 200)
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(x_vals, 2**x_vals, label='2^x')
ax.plot(x_vals, 3*2**x_vals, label='3·2^x (scala verticale)')
ax.plot(x_vals, 2**x_vals + 2, label='2^x+2 (trasla su, asintoto y=2)')
ax.plot(x_vals, 2**(-x_vals), label='2^(-x) (ribaltata)')
ax.axhline(y=0, color='gray', linewidth=0.5)
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_ylim(-1, 12)
plt.show()
````

## Esercizio svolto e commentato

**Problema:** confronta la crescita di $f(x)=2^x$ e $g(x)=x^{10}$ (una potenza fissa) per $x$ grande: quale cresce più velocemente?

**Ragionamento:**

**Intuizione iniziale (spesso sbagliata)**: $x^{10}$ ha un esponente grande, sembra "vincere" per piccoli valori di $x$. Verifichiamo numericamente prima di trarre conclusioni.

**Passo 1 — confrontiamo alcuni valori:**

| $x$ | $2^x$ | $x^{10}$ |
|---|---|---|
| $10$ | $1024$ | $10^{10}$ |
| $50$ | $\approx1.1\times10^{15}$ | $\approx9.8\times10^{16}$ |
| $100$ | $\approx1.3\times10^{30}$ | $10^{20}$ |
| $200$ | $\approx1.6\times10^{60}$ | $\approx1.0\times10^{23}$ |

**Passo 2 — osserviamo il cambio di tendenza.** Fino a circa $x=100$, $x^{10}$ sembra ancora competitiva o superiore. Ma tra $x=100$ e $x=200$, $2^x$ la supera **enormemente** ($10^{60}$ contro $10^{23}$): una volta superato un certo punto, l'esponenziale "stacca" definitivamente la potenza fissa, e il divario cresce sempre più.

**Conclusione**: **ogni** funzione esponenziale $a^x$ (con $a>1$) cresce, per $x$ sufficientemente grande, più velocemente di **qualunque** potenza fissa $x^k$ (per quanto grande sia $k$). Questo fatto — che formalizzeremo rigorosamente con i limiti nel Modulo 4 — è uno dei motivi per cui le funzioni esponenziali sono così importanti: descrivono una crescita "esplosiva", qualitativamente diversa da qualunque crescita polinomiale (notebook 3.11), non importa quanto alto sia il grado.

**Perché l'intuizione iniziale ("l'esponente grande vince") è ingannevole:** confonde il comportamento **iniziale** (per $x$ piccolo, dove effettivamente $x^{10}$ può essere più grande) con il comportamento **definitivo** (per $x$ grande, dove vince sempre l'esponenziale). È un errore comune: bisogna sempre chiedersi "per $x$ **abbastanza grande**", non fermarsi ai primi valori che si riescono a calcolare a mente.

````python
for x_val in [10, 50, 100, 200]:
    esponenziale = 2**x_val
    potenza = x_val**10
    print(f"x={x_val}: 2^x = {esponenziale:.4e}   x^10 = {potenza:.4e}   2^x maggiore? {esponenziale > potenza}")
````

## Riepilogo finale

- $f(x)=a^x$ ($a>0,a\ne1$): dominio $\mathbb{R}$, immagine $(0,+\infty)$, sempre iniettiva, strettamente monotona (crescente se $a>1$, decrescente se $0<a<1$), limitata solo da un lato.
- La base $e$ ha un ruolo speciale legato alla velocità di crescita, che formalizzeremo con le derivate (Modulo 5).
- Trasformazioni del grafico ($k\cdot a^x$, $a^x+c$, $a^{-x}$) permettono di costruire famiglie di modelli di crescita/decadimento a partire dalla forma base.
- Ogni esponenziale con base $>1$ cresce, definitivamente, più velocemente di qualunque potenza fissa $x^k$: una crescita qualitativamente più rapida di qualsiasi polinomio.
