# 3.12 — Funzioni razionali

**Prerequisiti:** 1.4 Frazioni algebriche, 3.11 Funzioni polinomiali, 2.7 Iperbole.


## Perché ci interessa

Una funzione razionale è il quoziente di due polinomi (notebook 1.4): $f(x)=\dfrac{P(x)}{Q(x)}$. A differenza dei polinomi (notebook 3.11), qui il dominio non è quasi mai tutto $\mathbb{R}$: il denominatore introduce esclusioni, e vicino a questi punti esclusi il grafico mostra un comportamento nuovo, gli **asintoti verticali** — lo stesso fenomeno già visto per la tangente (notebook 2.12) e per l'iperbole (notebook 2.7).

## Teoria, passo dopo passo

### Dominio

Il dominio di $f(x)=\dfrac{P(x)}{Q(x)}$ è $\mathbb{R}$ privato degli zeri di $Q(x)$ (notebook 1.4, 3.3): $D=\{x\in\mathbb{R} : Q(x)\ne0\}$.

### Asintoti verticali

Se $x_0$ è uno zero di $Q(x)$ che **non** è anche zero di $P(x)$ (cioè non si semplifica, notebook 1.4), il grafico di $f$ ha un **asintoto verticale** in $x=x_0$: avvicinandosi a $x_0$, $|f(x)|$ cresce senza limite (perché si divide per un numero sempre più vicino a $0$, notebook 0.3). Formalizzeremo rigorosamente questo comportamento con i limiti nel Modulo 4; per ora ci basta l'idea intuitiva, già incontrata con la tangente e l'iperbole.

**Attenzione**: se $x_0$ è zero **sia** di $P(x)$ sia di $Q(x)$, il fattore si semplifica (notebook 1.4), e il comportamento in $x_0$ può essere diverso (una "discontinuità eliminabile", che approfondiremo nel Modulo 5) — non un asintoto.

### Asintoti orizzontali: confronto dei gradi

Per $|x|$ grande, il comportamento di $f(x)=\dfrac{P(x)}{Q(x)}$ dipende dal confronto tra i gradi di $P$ e $Q$ (chiamiamoli $n$ e $m$):

- Se $n<m$: $f(x)\to0$ per $|x|$ grande (il denominatore "cresce più in fretta"): asintoto orizzontale $y=0$.
- Se $n=m$: $f(x)$ si avvicina al rapporto dei coefficienti dei termini di grado massimo, $\frac{a_n}{b_m}$: asintoto orizzontale $y=\frac{a_n}{b_m}$.
- Se $n>m$: $f(x)$ non ha asintoto orizzontale (cresce senza limite, come un polinomio, notebook 3.11); se $n=m+1$, ha invece un **asintoto obliquo** (una retta non orizzontale a cui il grafico si avvicina), che si trova con la divisione polinomiale.

Questo criterio è analogo, nello spirito, a quello già visto per l'iperbole (notebook 2.7): il rapporto tra parametri governa il comportamento "lontano dall'origine".

### Esempio guida: l'iperbole equilatera come funzione razionale

Ricordiamo dal notebook 2.7: $f(x)=\dfrac{k}{x}$ è un'iperbole equilatera scritta rispetto ai propri asintoti. Come funzione razionale: $P(x)=k$ (grado $0$), $Q(x)=x$ (grado $1$); poiché $n<m$, asintoto orizzontale $y=0$; e $x=0$ è l'unico zero del denominatore (non annulla il numeratore, se $k\ne0$): asintoto verticale $x=0$. Coerente con quanto già sappiamo dell'iperbole equilatera.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Esempio: f(x) = (2x^2-8)/(x^2-1), confronto dei gradi n=m=2
f = (2*x**2 - 8) / (x**2 - 1)
zeri_denominatore = sp.solve(sp.Eq(x**2 - 1, 0), x)
zeri_numeratore = sp.solve(sp.Eq(2*x**2 - 8, 0), x)
print("Zeri del denominatore (candidati asintoti verticali):", zeri_denominatore)
print("Zeri del numeratore:", zeri_numeratore)
print("Nessuna semplificazione: asintoti verticali confermati")

# Asintoto orizzontale: stesso grado, rapporto dei coefficienti principali
print("\nAsintoto orizzontale: y =", sp.Rational(2, 1))

# Grafico
x_vals = np.linspace(-5, 5, 1000)
y_vals = (2*x_vals**2 - 8) / (x_vals**2 - 1)
y_vals[np.abs(x_vals**2 - 1) < 0.05] = np.nan  # spezza vicino agli asintoti

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x_vals, y_vals, color='tab:blue')
ax.axvline(x=1, color='red', linestyle='--', label='asintoti verticali x=±1')
ax.axvline(x=-1, color='red', linestyle='--')
ax.axhline(y=2, color='green', linestyle='--', label='asintoto orizzontale y=2')
ax.set_ylim(-15, 15)
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_title("f(x) = (2x²-8)/(x²-1)")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** studia $f(x)=\dfrac{x^2-1}{x-1}$: dominio e comportamento in prossimità del punto escluso.

**Ragionamento:**

**Passo 1 — troviamo il dominio.** Il denominatore si annulla per $x=1$: $D=\mathbb{R}\setminus\{1\}$.

**Passo 2 — controlliamo se $x=1$ è anche zero del numeratore.** $x^2-1$ si annulla in $x=\pm1$ (differenza di quadrati, notebook 1.2): **sì**, $x=1$ annulla anche il numeratore!

**Passo 3 — semplifichiamo**, scomponendo il numeratore (notebook 1.3):

$$f(x) = \frac{(x-1)(x+1)}{x-1} = x+1 \qquad (x\ne1)$$

**Passo 4 — interpretiamo il risultato.** La funzione semplificata, $x+1$, è definita anche in $x=1$ (varrebbe $2$), ma la funzione **originale** $f$ non lo è (il dominio esclude $x=1$, indipendentemente dalla semplificazione — coerente con quanto visto nel notebook 3.1 sull'uguaglianza tra funzioni: $f$ e $g(x)=x+1$ **non sono la stessa funzione**, perché hanno domini diversi).

**Conclusione**: il grafico di $f$ coincide con la retta $y=x+1$ ovunque, **tranne** che manca il singolo punto $(1,2)$ — un piccolo "buco" nel grafico, non un asintoto verticale. Questo è l'esempio della "discontinuità eliminabile" menzionata nella teoria: il comportamento vicino a $x=1$ è perfettamente regolare (il grafico si avvicina in modo prevedibile a $(1,2)$ da entrambi i lati), solo che quel singolo punto non fa parte del dominio.

**Perché distinguere questo caso da un vero asintoto verticale è importante:** se avessimo concluso frettolosamente "$x=1$ annulla il denominatore, quindi c'è un asintoto verticale", avremmo sbagliato: un asintoto richiede che il valore assoluto della funzione cresca senza limite avvicinandosi al punto escluso (teoria), mentre qui la funzione si avvicina a un valore finito e ben preciso ($2$). Controllare sempre se il numeratore si annulla nello stesso punto è il passo che fa la differenza tra un buco e un vero asintoto.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = (x**2 - 1) / (x - 1)

semplificata = sp.simplify(f)
print("f semplificata:", semplificata)

# Valore a cui si avvicina f per x vicino a 1 (senza mai essere x=1)
for x_val in [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]:
    print(f"f({x_val}) = {f.subs(x, x_val)}")
````

## Riepilogo finale

- Il dominio di una funzione razionale $\frac{P(x)}{Q(x)}$ esclude gli zeri di $Q(x)$; se uno zero del denominatore è anche zero del numeratore, si semplifica e produce un "buco", non un asintoto.
- **Asintoto verticale** in $x_0$ se $Q(x_0)=0$ e $P(x_0)\ne0$: $|f(x)|$ cresce senza limite vicino a $x_0$.
- **Asintoto orizzontale**: confrontando i gradi $n$ (numeratore) e $m$ (denominatore) — $n<m$ dà $y=0$; $n=m$ dà $y=\frac{a_n}{b_m}$; $n>m$ nessun asintoto orizzontale (asintoto obliquo se $n=m+1$).
- L'iperbole equilatera (notebook 2.7) è l'esempio più semplice di funzione razionale con entrambi i tipi di asintoto.
