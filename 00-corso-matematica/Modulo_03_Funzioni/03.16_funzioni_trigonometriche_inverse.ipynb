# 3.16 — Funzioni trigonometriche inverse

**Prerequisiti:** 3.15 Funzioni trigonometriche, 3.6 Funzione inversa, 2.13 Coordinate polari.


## Perché ci interessa

Nel notebook 2.13 abbiamo usato $\theta=\arctan\left(\frac yx\right)$ senza definirla con precisione, promettendo di formalizzarla qui. Nel notebook 3.15 abbiamo trovato gli intervalli esatti su cui seno, coseno e tangente sono monotone (e quindi iniettive). Ora abbiamo tutto il necessario per costruire le loro funzioni inverse, seguendo esattamente il procedimento generale del notebook 3.6: restringere il dominio a un intervallo di iniettività, poi invertire.

## Teoria, passo dopo passo

### Arcoseno

Restringiamo $\sin$ all'intervallo $\left[-\frac\pi2,\frac\pi2\right]$ (notebook 3.15), dove è biiettiva verso $[-1,1]$. La sua inversa si chiama **arcoseno**:

$$\arcsin: [-1,1]\to\left[-\frac\pi2,\frac\pi2\right], \qquad \arcsin(y)=x \ \Longleftrightarrow\ \sin(x)=y,\ x\in\left[-\frac\pi2,\frac\pi2\right]$$

Il nome si legge "l'arco il cui seno è $y$". Dominio $[-1,1]$ (l'immagine del seno), immagine $\left[-\frac\pi2,\frac\pi2\right]$ (il dominio ristretto). Essendo l'inversa di una funzione crescente (notebook 3.15), anche $\arcsin$ è strettamente crescente (proprietà generale delle funzioni inverse, notebook 3.14).

### Arcocoseno

Restringiamo $\cos$ a $[0,\pi]$ (notebook 3.15), dove è biiettiva verso $[-1,1]$:

$$\arccos: [-1,1]\to[0,\pi], \qquad \arccos(y)=x \ \Longleftrightarrow\ \cos(x)=y,\ x\in[0,\pi]$$

Essendo l'inversa di una funzione decrescente, $\arccos$ è strettamente decrescente.

### Arcotangente

Restringiamo $\tan$ a $\left(-\frac\pi2,\frac\pi2\right)$ (notebook 3.15), dove è biiettiva verso tutto $\mathbb{R}$:

$$\arctan:\mathbb{R}\to\left(-\frac\pi2,\frac\pi2\right), \qquad \arctan(y)=x \ \Longleftrightarrow\ \tan(x)=y,\ x\in\left(-\frac\pi2,\frac\pi2\right)$$

Questo è esattamente il significato preciso di quanto usato "informalmente" nel notebook 2.13: l'arcotangente restituisce **sempre** un angolo tra $-\frac\pi2$ e $\frac\pi2$ (I o IV quadrante), da cui la necessità, notata allora, di correggere il risultato di $\pi$ per punti nel II o III quadrante — ora sappiamo esattamente **perché**: è una conseguenza diretta di come abbiamo dovuto restringere il dominio della tangente per renderla invertibile.

### Le identità delle funzioni inverse, applicate qui

Come per ogni funzione inversa (notebook 3.6): $\sin(\arcsin(y))=y$ per ogni $y\in[-1,1]$, e $\arcsin(\sin(x))=x$ **solo** per $x\in\left[-\frac\pi2,\frac\pi2\right]$ (non per ogni $x$! Se $x$ è fuori da quell'intervallo, $\arcsin(\sin(x))$ restituisce comunque un valore in $\left[-\frac\pi2,\frac\pi2\right]$, non $x$ stesso — coerente con la non iniettività del seno fuori da quell'intervallo, notebook 3.15).

**Attenzione a questo errore comune**: $\arcsin(\sin(\pi))\ne\pi$ (perché $\pi\notin\left[-\frac\pi2,\frac\pi2\right]$)! Calcoliamo correttamente: $\sin(\pi)=0$, e $\arcsin(0)=0$ (l'unico angolo in $\left[-\frac\pi2,\frac\pi2\right]$ con seno $0$). Quindi $\arcsin(\sin(\pi))=0\ne\pi$.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Verifica delle identità e dell'errore comune
print("sin(arcsin(0.5)) =", sp.sin(sp.asin(sp.Rational(1,2))))
print("arcsin(sin(pi/4)) =", sp.asin(sp.sin(sp.pi/4)), " (pi/4 è nell'intervallo, coincide)")
print("arcsin(sin(pi)) =", sp.asin(sp.sin(sp.pi)), " (pi NON è nell'intervallo, NON coincide con pi)")

# Grafici di arcoseno, arcocoseno, arcotangente
y_vals = np.linspace(-1, 1, 200)
x_vals_tan = np.linspace(-20, 20, 200)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].plot(y_vals, np.arcsin(y_vals))
axes[0].set_title("arcsin(y): dominio [-1,1], immagine [-π/2,π/2]")
axes[0].grid(True, alpha=0.3)

axes[1].plot(y_vals, np.arccos(y_vals))
axes[1].set_title("arccos(y): dominio [-1,1], immagine [0,π]")
axes[1].grid(True, alpha=0.3)

axes[2].plot(x_vals_tan, np.arctan(x_vals_tan))
axes[2].axhline(y=np.pi/2, color='red', linestyle='--', label='asintoto y=π/2')
axes[2].axhline(y=-np.pi/2, color='red', linestyle='--', label='asintoto y=-π/2')
axes[2].legend()
axes[2].set_title("arctan(x): dominio R, immagine (-π/2,π/2)")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** calcola $\cos(\arcsin(\frac35))$ senza usare una calcolatrice per gli angoli.

**Ragionamento:**

**Passo 1 — interpretiamo $\arcsin\left(\frac35\right)$.** È, per definizione, **l'angolo** $\alpha\in\left[-\frac\pi2,\frac\pi2\right]$ tale che $\sin\alpha=\frac35$. Non ci serve conoscere il valore numerico di $\alpha$: ci basta sapere che $\sin\alpha=\frac35$ e che $\alpha$ sta nell'intervallo dell'arcoseno.

**Passo 2 — usiamo l'identità fondamentale** (notebook 2.9): $\cos^2\alpha+\sin^2\alpha=1$, quindi:

$$\cos^2\alpha = 1-\left(\frac35\right)^2 = 1-\frac9{25}=\frac{16}{25}$$

**Passo 3 — determiniamo il segno di $\cos\alpha$.** Poiché $\alpha\in\left[-\frac\pi2,\frac\pi2\right]$ (l'immagine dell'arcoseno, teoria), siamo nel I o IV quadrante (o sull'asse), dove **il coseno è sempre $\ge0$** (notebook 2.9): scegliamo la radice positiva, senza ambiguità (a differenza dell'esercizio del notebook 2.9, dove servivano informazioni aggiuntive sul quadrante — qui la restrizione dell'arcoseno **ci garantisce già** il segno).

$$\cos\alpha = \sqrt{\frac{16}{25}} = \frac45$$

**Conclusione**: $\cos\left(\arcsin\frac35\right)=\frac45$.

**Perché conoscere l'immagine esatta dell'arcoseno (teoria) elimina l'ambiguità di segno che avremmo altrimenti:** nel notebook 2.9, per trovare $\cos\alpha$ da $\sin\alpha$ serviva sapere in quale quadrante ci si trovasse, altrimenti restavano due possibilità. Qui, la definizione stessa di $\arcsin$ **include** l'informazione sul quadrante (perché il suo codominio è ristretto a $\left[-\frac\pi2,\frac\pi2\right]$): è un esempio di come restringere accuratamente dominio/codominio, come fatto nel notebook 3.6 e 3.15, non sia solo un tecnicismo, ma renda le funzioni inverse strumenti calcolabili senza ambiguità.

````python
import sympy as sp

alpha = sp.asin(sp.Rational(3, 5))
risultato = sp.cos(alpha)
print("cos(arcsin(3/5)) =", sp.simplify(risultato))
````

## Riepilogo finale

- $\arcsin:[-1,1]\to\left[-\frac\pi2,\frac\pi2\right]$, $\arccos:[-1,1]\to[0,\pi]$, $\arctan:\mathbb{R}\to\left(-\frac\pi2,\frac\pi2\right)$: inverse delle restrizioni monotone individuate nel notebook 3.15.
- $f^{-1}(f(x))=x$ vale **solo** se $x$ appartiene all'intervallo di restrizione scelto; fuori da esso, il risultato è comunque un angolo nell'immagine della funzione inversa, non $x$ stesso.
- Conoscere l'immagine esatta di una funzione trigonometrica inversa elimina l'ambiguità di segno che si avrebbe altrimenti nel calcolare le altre funzioni trigonometriche a partire dal suo valore.
- Questo completa formalmente l'arcotangente già usata informalmente nel notebook 2.13 per le coordinate polari.
