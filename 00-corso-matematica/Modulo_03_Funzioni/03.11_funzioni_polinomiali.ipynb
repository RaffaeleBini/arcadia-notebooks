## Teoria, passo dopo passo

### Dominio e continuità

Ogni funzione polinomiale $f(x)=a_nx^n+\dots+a_1x+a_0$ ha **dominio $\mathbb{R}$**: non ci sono denominazioni, radici, o logaritmi (notebook 3.3) che possano restringerlo. Somme e prodotti di potenze di $x$ (con coefficienti reali) sono sempre calcolabili, per ogni $x$ reale.

### Il comportamento agli estremi: il termine di grado massimo comanda

Per $|x|$ molto grande, il comportamento di $f(x)=a_nx^n+\dots+a_0$ è dominato dal **termine di grado massimo**, $a_nx^n$: gli altri termini diventano trascurabili al confronto (un'idea intuitiva di limite, che formalizzeremo nel Modulo 4). Questo determina il comportamento "agli estremi" del grafico:

- Se $n$ è **pari** e $a_n>0$: il grafico "sale" sia a sinistra sia a destra (come $x^2$, notebook 2.5).
- Se $n$ è **pari** e $a_n<0$: "scende" da entrambi i lati (come $-x^2$).
- Se $n$ è **dispari** e $a_n>0$: "scende" a sinistra, "sale" a destra (come $x^3$).
- Se $n$ è **dispari** e $a_n<0$: comportamento opposto.

### Limitatezza

Conseguenza diretta di quanto sopra: **nessun polinomio non costante è limitato** (notebook 3.10), né superiormente né inferiormente, se ha grado dispari; se ha grado pari, è limitato solo da un lato (quello opposto al segno di $a_n$) — esattamente il comportamento di $x^2$ visto nel notebook 3.10. Solo i polinomi costanti ($n=0$) sono limitati in entrambe le direzioni (banalmente, l'immagine è un solo punto).

### Parità

Un polinomio con **soli** esponenti pari è una funzione pari (notebook 3.8); con soli esponenti dispari, è dispari. Un polinomio con **entrambi i tipi di esponenti** (come $x^3+x^2$) non è né pari né dispari in generale — l'esempio di questo caso lo avevamo già incontrato proprio nel notebook 3.8.

### Numero di intersezioni con l'asse $x$ (radici)

Un polinomio di grado $n$ ha **al massimo $n$ radici reali** (soluzioni di $f(x)=0$) — un fatto che generalizza quanto già visto per l'equazione di secondo grado (notebook 1.6, al massimo 2 soluzioni) e useremo spesso. Per trovarle, servono le tecniche di scomposizione del notebook 1.3, oppure (per gradi $\ge3$) tecniche più avanzate come il teorema di Ruffini, che non trattiamo in questo corso ma che potrai approfondire in seguito.

### Monotonia locale

A differenza delle rette (sempre monotone su tutto $\mathbb{R}$, notebook 2.2), i polinomi di grado $\ge2$ tipicamente alternano tratti crescenti e decrescenti (come già visto per $x^2$, notebook 3.9). Il numero massimo di "cambi di direzione" di un polinomio di grado $n$ è $n-1$: per esempio un polinomio di terzo grado può avere al più 2 cambi di direzione (un massimo locale e un minimo locale). Studieremo con precisione questi punti nel Modulo 5, con le derivate.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Confrontiamo il comportamento agli estremi per diversi gradi e segni
fig, axes = plt.subplots(2, 2, figsize=(11, 9))
x_vals = np.linspace(-3, 3, 200)

casi = [
    (x_vals**2, "x²  (n pari, a>0): sale da entrambi i lati"),
    (-x_vals**2, "-x²  (n pari, a<0): scende da entrambi i lati"),
    (x_vals**3, "x³  (n dispari, a>0): scende a sx, sale a dx"),
    (-x_vals**3, "-x³  (n dispari, a<0): comportamento opposto"),
]

for ax, (y_vals, titolo) in zip(axes.flat, casi):
    ax.plot(x_vals, y_vals)
    ax.axhline(y=0, color='gray', linewidth=0.5)
    ax.set_title(titolo, fontsize=10)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Verifica del numero massimo di radici reali
f = x**3 - 3*x + 1  # polinomio di grado 3
radici = sp.solve(sp.Eq(f, 0), x)
print(f"Radici di {f}: {radici}  (al massimo 3, come previsto)")
````

## Esercizio svolto e commentato

**Problema:** studia $f(x)=x^4-4x^2$: dominio, parità, comportamento agli estremi, e le sue radici.

**Ragionamento:**

**Passo 1 — dominio**: $\mathbb{R}$ (ogni polinomio, teoria).

**Passo 2 — parità**: tutti gli esponenti presenti ($4$ e $2$) sono pari (notebook 3.8): $f(-x)=(-x)^4-4(-x)^2=x^4-4x^2=f(x)$. **Funzione pari**, grafico simmetrico rispetto all'asse $y$.

**Passo 3 — comportamento agli estremi**: grado $n=4$ (pari), coefficiente del termine di grado massimo $a_4=1>0$: il grafico **sale da entrambi i lati** (teoria).

**Passo 4 — troviamo le radici**, risolvendo $x^4-4x^2=0$. Raccogliamo $x^2$ (notebook 1.3):

$$x^2(x^2-4)=0$$

Per la legge di annullamento del prodotto (notebook 1.6): $x^2=0$ oppure $x^2-4=0$. Dalla prima: $x=0$ (radice doppia, perché viene da $x^2$). Dalla seconda (differenza di quadrati, notebook 1.2): $x=\pm2$.

Le radici sono $x=-2,0,2$ (tre radici distinte, ma contando la molteplicità di $x=0$, siamo a 4 "radici" totali, coerente con il grado 4 — un dettaglio sulla molteplicità che approfondiremo più avanti nel corso).

**Passo 5 — sintesi del comportamento**: essendo pari (simmetrico rispetto all'asse $y$) e con questi tre punti di intersezione, il grafico scende da $+\infty$ (a sinistra), attraversa l'asse in $x=-2$, scende ancora fino a un minimo locale, risale toccando (senza attraversare, essendo radice doppia) l'asse in $x=0$, poi ridiscende simmetricamente e risale di nuovo verso $+\infty$ dopo $x=2$: una "doppia U".

**Perché il fatto che $f$ sia pari dimezza il lavoro necessario per capire il grafico:** avendo stabilito la parità al passo 2, sappiamo che il comportamento per $x<0$ è automaticamente lo specchio di quello per $x>0$ (notebook 3.8). Potremmo quindi studiare $f$ solo su $[0,+\infty)$ e dedurre il resto per simmetria, dimezzando effettivamente l'analisi da svolgere.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)
f = x**4 - 4*x**2

radici = sp.solve(sp.Eq(f, 0), x)
print("Radici:", radici)

print("f è pari?", sp.simplify(f.subs(x, -x) - f) == 0)

x_vals = np.linspace(-2.5, 2.5, 300)
y_vals = x_vals**4 - 4*x_vals**2
plt.figure(figsize=(7, 5))
plt.plot(x_vals, y_vals)
plt.axhline(y=0, color='gray', linewidth=0.5)
plt.plot([-2, 0, 2], [0, 0, 0], 'ro')
plt.title("f(x) = x⁴ - 4x²: pari, radici in -2, 0 (doppia), 2")
plt.grid(True, alpha=0.3)
plt.show()
````

## Riepilogo finale

- Ogni funzione polinomiale ha dominio $\mathbb{R}$; il comportamento "agli estremi" (per $|x|$ grande) è determinato dal termine di grado massimo.
- Grado pari con $a_n>0$: sale da entrambi i lati; grado pari con $a_n<0$: scende da entrambi; grado dispari: comportamenti opposti a sinistra e destra.
- Un polinomio non costante non è mai limitato in entrambe le direzioni (al più da un lato, se il grado è pari).
- Un polinomio di grado $n$ ha al massimo $n$ radici reali; la parità dipende dalla presenza esclusiva di esponenti pari o dispari.
