# 3.3 — Dominio

**Prerequisiti:** 3.2 Funzioni reali, 1.4 Frazioni algebriche, 0.6 Frazioni potenze e radici, 1.12 Logaritmi.


## Perché ci interessa

Nel notebook 3.2 abbiamo visto che il dominio di una funzione può essere tutto $\mathbb{R}$, oppure un sottoinsieme più piccolo (come $[0,+\infty)$ per $\sqrt{x}$, o $(0,+\infty)$ per $\log x$). Quando una funzione è data da una formula "senza altre precisazioni", il **dominio naturale** è l'insieme di tutti gli $x$ per cui la formula ha senso.

Abbiamo già incontrato singolarmente ciascuna delle regole che servono: denominatori diversi da zero (notebook 1.4), radicandi non negativi per radici pari (notebook 0.6), argomenti positivi per i logaritmi (notebook 1.12). In questo notebook le raccogliamo in un procedimento sistematico, utile anche quando più condizioni si combinano nella stessa espressione.

## Teoria, passo dopo passo

### Le condizioni tipiche da imporre

Data un'espressione $f(x)$, per trovarne il dominio si cercano tutte le operazioni "pericolose" al suo interno, e si impone che siano ben definite:

1. **Denominatore**: se $f(x)$ contiene $\frac{1}{g(x)}$, serve $g(x)\ne0$ (notebook 1.4).
2. **Radice di indice pari** (quadrata, quarta, ...): se $f(x)$ contiene $\sqrt[2k]{g(x)}$, serve $g(x)\ge0$ (notebook 0.6 — le radici di indice dispari, invece, non impongono nessuna condizione: sono definite anche per argomenti negativi, es. $\sqrt[3]{-8}=-2$).
3. **Logaritmo**: se $f(x)$ contiene $\log_a(g(x))$, serve $g(x)>0$ (notebook 1.12).
4. **Base di una potenza con esponente non intero**: se $f(x)$ contiene $g(x)^{h(x)}$ con $h(x)$ non intero, in genere si richiede $g(x)>0$ (coerente con l'estensione delle potenze del notebook 1.10, che richiedeva base positiva).

Quando un'espressione contiene **più** di queste condizioni contemporaneamente, il dominio è l'**intersezione** di tutte (notebook 0.13): un valore di $x$ deve soddisfarle **tutte** per essere ammesso.

### Procedimento sistematico

1. Identifica ogni "punto critico" nella formula (ogni denominatore, radice pari, logaritmo...).
2. Scrivi la condizione richiesta da ciascuno, come equazione o disequazione (notebook 1.5-1.9).
3. Risolvi ciascuna condizione separatamente.
4. Interseca tutte le soluzioni (notebook 0.13): il risultato è il dominio.

### Un esempio guida

$$f(x) = \frac{\sqrt{x+2}}{\log(x-1)}$$

Qui ci sono **tre** condizioni: il radicando della radice quadrata non negativo, l'argomento del logaritmo positivo, e il denominatore (l'intero logaritmo) diverso da zero.

- Radice: $x+2\ge0 \Rightarrow x\ge-2$.
- Logaritmo (argomento positivo): $x-1>0 \Rightarrow x>1$.
- Denominatore (il logaritmo stesso $\ne0$): $\log(x-1)\ne0 \Rightarrow x-1\ne1 \Rightarrow x\ne2$ (perché $\log_a(1)=0$ per qualunque base, notebook 1.12).

Intersecando: $x\ge-2$ **e** $x>1$ **e** $x\ne2$. Le prime due si intersecano in $x>1$ (più restrittiva); togliendo il punto $x=2$:

$$D = (1,2)\cup(2,+\infty)$$

````python
import sympy as sp

x = sp.symbols('x', real=True)

# SymPy può calcolare il dominio automaticamente con calculate_domain (via continuous_domain)
from sympy.calculus.util import continuous_domain

f = sp.sqrt(x + 2) / sp.log(x - 1)
dominio = continuous_domain(f, x, sp.S.Reals)
print("Dominio calcolato da SymPy:", dominio)

# Verifichiamo passo passo le singole condizioni
cond_radice = sp.solve(x + 2 >= 0, x)
cond_log_argomento = sp.solve(x - 1 > 0, x)
cond_denominatore = sp.solve(sp.Eq(sp.log(x - 1), 0), x)  # valori da ESCLUDERE
print("\nCondizione radice (x+2>=0):", cond_radice)
print("Condizione log argomento (x-1>0):", cond_log_argomento)
print("Valori da escludere (dove il log si annulla):", cond_denominatore)
````

## Esercizio svolto e commentato

**Problema:** trova il dominio di $f(x) = \dfrac{1}{\sqrt{4-x^2}}$.

**Ragionamento:**

**Passo 1 — identifichiamo le condizioni.** Qui c'è una radice quadrata (indice pari, richiede radicando $\ge0$) **dentro** un denominatore (richiede l'intera espressione $\ne0$). Combinando le due: il radicando deve essere **strettamente positivo** (non solo $\ge0$): se fosse $=0$, la radice varrebbe $0$, e avremmo comunque una divisione per zero.

$$4-x^2 > 0$$

**Passo 2 — risolviamo la disequazione** (notebook 1.9, disequazione di secondo grado): scomponiamo $4-x^2=(2-x)(2+x)$ (differenza di quadrati, notebook 1.2). Studiamo il segno:

| Intervallo | $2-x$ | $2+x$ | $(2-x)(2+x)$ |
|---|---|---|---|
| $x<-2$ | $+$ | $-$ | $-$ |
| $-2<x<2$ | $+$ | $+$ | $+$ |
| $x>2$ | $-$ | $+$ | $-$ |

Vogliamo dove il prodotto è **positivo** (strettamente): $-2<x<2$.

**Passo 3 — scriviamo il dominio:**

$$D=(-2,2)$$

(intervallo aperto: gli estremi $\pm2$ sono esclusi, perché lì $4-x^2=0$, non $>0$).

**Perché la condizione è "$>0$" e non "$\ge0$" come per una radice quadrata isolata:** se la radice fosse da sola (non al denominatore), basterebbe $4-x^2\ge0$ (radicando non negativo, notebook 0.6), includendo anche gli estremi dove la radice vale $0$. Ma qui la radice è al **denominatore**: dividere per $0$ non è mai ammesso (notebook 0.3), quindi dobbiamo escludere anche i punti dove il radicando si annulla, restringendo la condizione a "$>0$" stretto. Combinare più condizioni richiede sempre di prestare attenzione a come interagiscono tra loro, non solo applicarle separatamente.

````python
import sympy as sp
from sympy.calculus.util import continuous_domain

x = sp.symbols('x', real=True)
f = 1 / sp.sqrt(4 - x**2)
dominio = continuous_domain(f, x, sp.S.Reals)
print("Dominio:", dominio)
````

3. Trova il dominio di $f(x)=\log(x^2-4)$.
4. *(più impegnativo)* Trova il dominio di $f(x)=\dfrac{\sqrt{x}}{\log(3-x)}$ (tre condizioni da combinare).

<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Denominatore $\ne0$: $x^2-9\ne0 \Rightarrow x\ne\pm3$. Dominio: $\mathbb{R}\setminus\{-3,3\}$.

**2.** Radicando $\ge0$: $2x-6\ge0 \Rightarrow x\ge3$. Dominio: $[3,+\infty)$.

**3.** Argomento $>0$: $x^2-4>0$, cioè $(x-2)(x+2)>0$, vero per $x<-2$ oppure $x>2$. Dominio: $(-\infty,-2)\cup(2,+\infty)$.

**4.** Radice: $x\ge0$. Logaritmo (argomento positivo): $3-x>0 \Rightarrow x<3$. Denominatore (log $\ne0$): $3-x\ne1 \Rightarrow x\ne2$. Intersecando $x\ge0$ e $x<3$ e $x\ne2$: $D=[0,2)\cup(2,3)$.

Verifica computazionale:
```python

print("Esercizio 1:", continuous_domain((x+1)/(x**2-9), x, sp.S.Reals))
print("Esercizio 2:", continuous_domain(sp.sqrt(2*x-6), x, sp.S.Reals))
print("Esercizio 3:", continuous_domain(sp.log(x**2-4), x, sp.S.Reals))
print("Esercizio 4:", continuous_domain(sp.sqrt(x)/sp.log(3-x), x, sp.S.Reals))
```

## Riepilogo finale

- Il dominio naturale di una funzione è l'insieme degli $x$ per cui la formula ha senso: denominatori $\ne0$ (notebook 1.4), radicandi di indice pari $\ge0$ (notebook 0.6), argomenti di logaritmi $>0$ (notebook 1.12).
- Con più condizioni contemporaneamente, il dominio è l'**intersezione** di tutte (notebook 0.13).
- Attenzione a come le condizioni interagiscono: una radice al denominatore richiede il radicando strettamente positivo (non solo $\ge0$), perché combina due vincoli.
- Il procedimento è sempre lo stesso: identificare i punti critici, scrivere e risolvere ogni condizione, intersecare i risultati.
