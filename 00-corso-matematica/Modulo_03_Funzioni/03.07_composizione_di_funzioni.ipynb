# 3.7 — Composizione di funzioni

**Prerequisiti:** 3.2 Funzioni reali, 3.6 Funzione inversa.


## Perché ci interessa

Nel notebook 3.6 abbiamo scritto $f^{-1}(f(x))=x$: applicare prima $f$ e poi $f^{-1}$. Questa operazione — "applicare una funzione al risultato di un'altra" — è così comune e utile da meritare un nome proprio: **composizione**. La useremo per costruire funzioni complesse a partire da funzioni semplici (e viceversa, per "smontare" funzioni complesse in pezzi più semplici da studiare).

## Teoria, passo dopo passo

### Definizione

Date due funzioni $f$ e $g$, la **composizione** $f\circ g$ ("f composto g") è la funzione definita da:

$$(f\circ g)(x) = f(g(x))$$

Si legge: prima si applica $g$ a $x$, poi si applica $f$ al risultato. **L'ordine conta**, e si legge da destra verso sinistra nel simbolo $f\circ g$: $g$ agisce per prima.

Esempio: $f(x)=x^2$, $g(x)=x+1$. Allora $(f\circ g)(x)=f(g(x))=f(x+1)=(x+1)^2$. Ma $(g\circ f)(x)=g(f(x))=g(x^2)=x^2+1$. Sono **funzioni diverse**: $(1+x)^2$ non è, in generale, uguale a $x^2+1$ (verificabile sviluppando il quadrato, notebook 1.2: $(x+1)^2=x^2+2x+1\ne x^2+1$, a meno che $x=0$). La composizione, in generale, **non è commutativa**.

### Il dominio della composizione

Per calcolare $(f\circ g)(x)$, servono due condizioni: $x$ deve stare nel dominio di $g$ (altrimenti $g(x)$ non è definito), e $g(x)$ deve stare nel dominio di $f$ (altrimenti $f(g(x))$ non è definito). Il dominio di $f\circ g$ è quindi:

$$D_{f\circ g} = \{x\in D_g : g(x)\in D_f\}$$

Questo procedimento — verificare **entrambe** le condizioni — è concettualmente simile alla ricerca del dominio di un'espressione composta (notebook 3.3): ogni "livello" della composizione impone la propria condizione.

### La funzione identità

La **funzione identità**, $\text{id}(x)=x$, è l'elemento "neutro" della composizione (analogo a $0$ per l'addizione o $1$ per la moltiplicazione, notebook 0.5):

$$(f\circ \text{id})(x) = f(\text{id}(x)) = f(x) \qquad (\text{id}\circ f)(x) = \text{id}(f(x)) = f(x)$$

Con questo linguaggio, le identità del notebook 3.6 si scrivono in modo compatto:

$$f^{-1}\circ f = \text{id}_D \qquad f\circ f^{-1} = \text{id}_B$$

("comporre $f$ con la sua inversa, in un ordine o nell'altro, dà sempre l'identità" — sul dominio appropriato in ciascun caso).

### Composizione e iniettività/suriettività

Un fatto utile: se $f$ e $g$ sono entrambe iniettive (notebook 3.5), anche $f\circ g$ è iniettiva. Intuizione: se $(f\circ g)(x_1)=(f\circ g)(x_2)$, cioè $f(g(x_1))=f(g(x_2))$, per l'iniettività di $f$ otteniamo $g(x_1)=g(x_2)$; per l'iniettività di $g$, $x_1=x_2$. Un ragionamento a catena, tipico di molte dimostrazioni su funzioni composte.

````python
import sympy as sp

x = sp.symbols('x', real=True)

f = x**2
g = x + 1

f_composto_g = f.subs(x, g)
g_composto_f = g.subs(x, f)
print("(f∘g)(x) = f(g(x)) =", sp.expand(f_composto_g))
print("(g∘f)(x) = g(f(x)) =", sp.expand(g_composto_f))
print("Sono uguali?", sp.expand(f_composto_g) == sp.expand(g_composto_f))

# Verifica della funzione identità come elemento neutro
identita = x
print("\n(f∘id)(x) =", f.subs(x, identita), " (deve coincidere con f(x))")

# Verifica che f∘f^-1 = identità, per f(x)=2x+3
f2 = 2*x + 3
f2_inversa = (x - 3) / 2
composizione = sp.simplify(f2.subs(x, f2_inversa))
print("\n(f∘f⁻¹)(x) =", composizione, " (deve essere x)")
````

## Esercizio svolto e commentato

**Problema:** date $f(x)=\sqrt{x}$ e $g(x)=x-4$, trova $(f\circ g)(x)$ e il suo dominio.

**Ragionamento:**

**Passo 1 — scriviamo la composizione:**

$$(f\circ g)(x) = f(g(x)) = f(x-4) = \sqrt{x-4}$$

**Passo 2 — troviamo il dominio, seguendo la definizione della teoria.** Serve prima $x\in D_g$: $g(x)=x-4$ è definita ovunque, $D_g=\mathbb{R}$, nessuna restrizione da qui. Poi serve $g(x)\in D_f$: $D_f=[0,+\infty)$ (dominio della radice quadrata, notebook 0.6), quindi:

$$g(x)\ge0 \quad\Rightarrow\quad x-4\ge0 \quad\Rightarrow\quad x\ge4$$

Il dominio di $f\circ g$ è $[4,+\infty)$.

**Confronto istruttivo — calcoliamo anche $(g\circ f)(x)$:**

$$(g\circ f)(x) = g(f(x)) = g(\sqrt{x}) = \sqrt{x}-4$$

Qui il dominio richiede solo $x\in D_f=[0,+\infty)$ (il risultato $\sqrt x-4$ è sempre un numero reale ben definito, qualunque sia il segno, perché $g$ è definita su tutto $\mathbb{R}$): $D_{g\circ f}=[0,+\infty)$, diverso dal dominio di $f\circ g$ trovato sopra!

**Perché i domini di $f\circ g$ e $g\circ f$ possono essere diversi, anche quando entrambe le composizioni "hanno senso" come formula:** il dominio dipende da **quale funzione agisce per prima**. In $f\circ g$, prima si applica $g$ (senza restrizioni), poi bisogna che il risultato sia $\ge0$ per $f=\sqrt{\cdot}$: da qui la restrizione $x\ge4$. In $g\circ f$, prima si applica $f=\sqrt{\cdot}$ (che già richiede $x\ge0$), poi $g$ non aggiunge altre condizioni. Le due composizioni sono strutturalmente diverse, non solo nella formula finale ma anche in *dove* nasce ciascuna restrizione.

````python
import sympy as sp
from sympy.calculus.util import continuous_domain

x = sp.symbols('x', real=True)

f_composto_g = sp.sqrt(x - 4)
dominio_fg = continuous_domain(f_composto_g, x, sp.S.Reals)
print("Dominio di f∘g:", dominio_fg)

g_composto_f = sp.sqrt(x) - 4
dominio_gf = continuous_domain(g_composto_f, x, sp.S.Reals)
print("Dominio di g∘f:", dominio_gf)
````

## Riepilogo finale

- $(f\circ g)(x)=f(g(x))$: prima si applica $g$, poi $f$. In generale $f\circ g \ne g\circ f$.
- Il dominio di $f\circ g$ richiede $x\in D_g$ **e** $g(x)\in D_f$: entrambe le condizioni, in questo ordine.
- La funzione identità $\text{id}(x)=x$ è l'elemento neutro della composizione; le identità del notebook 3.6 si scrivono $f^{-1}\circ f=\text{id}$ e $f\circ f^{-1}=\text{id}$.
- La composizione di due funzioni iniettive è iniettiva (dimostrabile con un ragionamento a catena).
