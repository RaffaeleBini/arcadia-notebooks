# 3.10 — Funzioni limitate

**Prerequisiti:** 3.4 Codominio e immagine, 0.14 Intervalli.


## Perché ci interessa

Nel notebook 2.12 abbiamo osservato che $\sin\theta$ e $\cos\theta$ restano sempre tra $-1$ e $1$, mentre $\tan\theta$ può diventare arbitrariamente grande vicino ai suoi asintoti. Questa differenza — "l'immagine sta dentro un intervallo finito" oppure no — è la proprietà di **limitatezza**, che studiamo ora in generale. È una proprietà semplice da enunciare ma con conseguenze profonde, che ritroveremo nel Modulo 5 parlando di massimi e minimi.

## Teoria, passo dopo passo

### Limitata superiormente e inferiormente

Una funzione $f:D\to\mathbb{R}$ è **limitata superiormente** se esiste un numero $M$ tale che $f(x)\le M$ per **ogni** $x\in D$ (il valore $M$ si chiama **maggiorante**). È **limitata inferiormente** se esiste $m$ tale che $f(x)\ge m$ per ogni $x\in D$ (**minorante**). Usando i quantificatori (notebook 0.16):

$$\text{limitata superiormente}: \quad \exists M\in\mathbb{R},\ \forall x\in D,\ f(x)\le M$$

Una funzione **limitata** (senza altre precisazioni) è limitata sia superiormente sia inferiormente: la sua immagine (notebook 3.4) è contenuta in un intervallo limitato $[m,M]$.

### Esempi già noti

- $\sin x$ e $\cos x$ (notebook 2.12) sono limitate: immagine $[-1,1]$, quindi $M=1$, $m=-1$ funzionano.
- $\tan x$ **non** è limitata: vicino ai suoi asintoti verticali (notebook 2.12), assume valori grandi a piacere, sia positivi sia negativi.
- $f(x)=x^2$ (notebook 2.5) è limitata **inferiormente** (da $m=0$, dato che $x^2\ge0$ sempre) ma **non superiormente** (per $|x|$ grande, $x^2$ cresce senza limite).
- $f(x)=a^x$ con $a>1$ (notebook 1.11) è limitata inferiormente (da $m=0$, l'esponenziale è sempre positiva) ma non superiormente.

### Maggioranti e minoranti non sono unici

Se $M$ è un maggiorante, anche **ogni** numero più grande di $M$ lo è (la condizione $f(x)\le M$ implica automaticamente $f(x)\le M'$ per ogni $M'>M$, per transitività dell'ordine). Per esempio, per $\sin x$: $1$, $2$, $100$ sono tutti maggioranti validi. Il maggiorante **più piccolo possibile**, se esiste, si chiama **estremo superiore** ($\sup$): per $\sin x$, $\sup=1$ (ed è anche il massimo, perché viene effettivamente raggiunto, es. in $\theta=\frac{\pi}{2}$). Non approfondiamo qui la teoria completa di estremo superiore/inferiore, che richiede la proprietà di completezza di $\mathbb{R}$ — la vedremo con rigore nel Modulo 5, dove sarà uno strumento fondamentale.

### Limitatezza e composizione/somma

Alcune proprietà utili, dirette conseguenze delle definizioni:

- La somma di due funzioni limitate è limitata (se $|f(x)|\le M_1$ e $|g(x)|\le M_2$, allora $|f(x)+g(x)|\le M_1+M_2$, per la disuguaglianza triangolare, notebook 0.8).
- Il prodotto di una funzione limitata per una funzione limitata è limitato, con un ragionamento analogo.
- Una funzione **continua su un intervallo chiuso e limitato** (concetto che formalizzeremo nel Modulo 5) è sempre limitata: un risultato importante, il teorema di Weierstrass, che vedremo con tutti i dettagli più avanti.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from sympy.calculus.util import function_range

x = sp.symbols('x', real=True)

funzioni = {
    "sin(x)": sp.sin(x),
    "x^2": x**2,
    "2^x": 2**x,
}

for nome, f in funzioni.items():
    immagine = function_range(f, x, sp.S.Reals)
    print(f"{nome}: immagine = {immagine}")

# Grafico che mostra visivamente la limitatezza (o meno)
theta = np.linspace(-2*np.pi, 2*np.pi, 500)
x_vals = np.linspace(-3, 3, 500)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(theta, np.sin(theta))
axes[0].axhline(y=1, color='red', linestyle='--')
axes[0].axhline(y=-1, color='red', linestyle='--')
axes[0].set_title("sin(x): limitata, [-1,1]")
axes[0].grid(True, alpha=0.3)

axes[1].plot(x_vals, x_vals**2)
axes[1].axhline(y=0, color='red', linestyle='--', label='limitata inferiormente')
axes[1].set_title("x²: limitata SOLO inferiormente")
axes[1].legend(); axes[1].grid(True, alpha=0.3)

axes[2].plot(x_vals, 2.0**x_vals)
axes[2].axhline(y=0, color='red', linestyle='--')
axes[2].set_title("2^x: limitata SOLO inferiormente")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $f(x)=\dfrac{x}{x^2+1}$ (dominio $\mathbb{R}$) è limitata.

**Ragionamento:**

**Passo 1 — cerchiamo l'immagine**, con la strategia del notebook 3.4: impostiamo $\frac{x}{x^2+1}=y$ e ci chiediamo per quali $y$ l'equazione ha soluzione $x$ reale.

$$x = y(x^2+1) \quad\Rightarrow\quad yx^2-x+y=0$$

**Passo 2 — trattiamo come equazione di secondo grado in $x$** (notebook 1.6). Se $y=0$: l'equazione diventa $-x=0$, cioè $x=0$, soluzione valida. Se $y\ne0$, il discriminante deve essere $\ge0$:

$$\Delta = 1-4y^2 \ge 0 \quad\Rightarrow\quad y^2\le\frac14 \quad\Rightarrow\quad -\frac12\le y\le\frac12$$

**Passo 3 — includiamo il caso $y=0$** (già trovato) in questo intervallo (che lo contiene già): l'immagine è $\left[-\frac12,\frac12\right]$.

**Conclusione**: l'immagine è contenuta nell'intervallo limitato $\left[-\frac12,\frac12\right]$: $f$ è **limitata**, con $M=\frac12$ (maggiorante, anzi il massimo, raggiunto per $x=1$: $f(1)=\frac{1}{2}$) e $m=-\frac12$ (minimo, raggiunto per $x=-1$).

**Perché il metodo del notebook 3.4 (trasformare in equazione di secondo grado, imporre $\Delta\ge0$) è particolarmente efficace per stabilire la limitatezza:** trovando l'**intero** intervallo dell'immagine in un colpo solo, otteniamo automaticamente sia l'informazione sulla limitatezza sia i valori esatti di massimo e minimo (quando raggiunti) — molto più informativo che verificare la limitatezza "a tentativi" su singoli punti, che potrebbe solo suggerire un limite senza mai confermarlo con certezza.

````python
import sympy as sp
from sympy.calculus.util import function_range

x = sp.symbols('x', real=True)
f = x / (x**2 + 1)
immagine = function_range(f, x, sp.S.Reals)
print("Immagine:", immagine)

print("f(1) =", f.subs(x, 1))
print("f(-1) =", f.subs(x, -1))
````

## Riepilogo finale

- $f$ è **limitata superiormente** se $f(x)\le M$ per un qualche $M$, per ogni $x$ nel dominio; **limitata inferiormente** analogamente con $\ge m$; **limitata** se entrambe.
- Maggioranti e minoranti non sono unici: se $M$ funziona, anche ogni numero più grande funziona.
- Trovare l'immagine (notebook 3.4) è spesso il modo più diretto per stabilire la limitatezza: se l'immagine è contenuta in un intervallo limitato, la funzione è limitata.
- Somma e prodotto di funzioni limitate restano limitate; questa proprietà è il primo passo verso il teorema di Weierstrass, che vedremo nel Modulo 5.
