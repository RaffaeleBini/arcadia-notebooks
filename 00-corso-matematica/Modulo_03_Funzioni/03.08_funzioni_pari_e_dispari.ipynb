# 3.8 — Funzioni pari e dispari

**Prerequisiti:** 3.2 Funzioni reali, 2.10 Identità trigonometriche.


## Perché ci interessa

Nel notebook 2.10 abbiamo osservato, senza formalizzarlo, che $\cos(-\theta)=\cos\theta$ e $\sin(-\theta)=-\sin\theta$. Queste sono manifestazioni di una proprietà generale delle funzioni: la **simmetria** rispetto all'asse $y$ o all'origine. Riconoscere questa proprietà in anticipo fa risparmiare lavoro: basta studiare una funzione "pari" o "dispari" per $x\ge0$, e il comportamento per $x<0$ è automaticamente determinato.

## Teoria, passo dopo passo

### Funzioni pari

Una funzione $f$ (con dominio simmetrico rispetto a $0$, cioè $x\in D \Rightarrow -x\in D$) è **pari** se:

$$f(-x)=f(x) \qquad \text{per ogni } x\in D$$

Il grafico di una funzione pari è **simmetrico rispetto all'asse $y$**: il punto $(x,f(x))$ ha sempre il suo "gemello" $(-x,f(x))$, alla stessa altezza. Esempio: $f(x)=x^2$ (notebook 2.5): $f(-x)=(-x)^2=x^2=f(x)$. Anche $\cos\theta$ è pari (notebook 2.10).

### Funzioni dispari

Una funzione $f$ (stesso requisito sul dominio) è **dispari** se:

$$f(-x)=-f(x) \qquad \text{per ogni } x\in D$$

Il grafico di una funzione dispari è **simmetrico rispetto all'origine** (una rotazione di $180°$ del grafico coincide con sé stesso): il punto $(x,f(x))$ ha come "gemello" $(-x,-f(x))$. Esempio: $f(x)=x^3$ (notebook 3.6, esercizio 2): $f(-x)=(-x)^3=-x^3=-f(x)$. Anche $\sin\theta$ e $\tan\theta$ sono dispari (notebook 2.10, 2.12).

### La maggior parte delle funzioni non è né pari né dispari

Attenzione: "pari" e "dispari" **non sono le uniche due possibilità** (a differenza di come si comportano i numeri interi, notebook 0.2, dove ogni intero è pari o dispari, mai nessuno dei due — da cui probabilmente nasce la tentazione di pensare lo stesso per le funzioni). Per esempio $f(x)=x^2+x$: $f(-x)=x^2-x$, che non è né uguale a $f(x)=x^2+x$ (non è pari) né uguale a $-f(x)=-x^2-x$ (non è dispari). La maggior parte delle funzioni "generiche" non ha nessuna delle due simmetrie.

### Come riconoscere pari/dispari da una formula

Alcune regole pratiche, dirette conseguenze delle definizioni:

- Un **polinomio con soli esponenti pari** (incluso il termine costante, "esponente $0$") è una funzione pari: es. $3x^4-2x^2+5$.
- Un **polinomio con soli esponenti dispari** (nessun termine costante) è una funzione dispari: es. $x^5-3x^3+2x$.
- Somma di due funzioni pari è pari; somma di due dispari è dispari (verificabile direttamente dalla definizione, come nell'esercizio del notebook 0.1 su "pari+pari=pari").
- **Prodotto**: pari$\times$pari = pari; dispari$\times$dispari = pari; pari$\times$dispari = dispari — esattamente la stessa struttura della regola dei segni (notebook 0.2)! Non è un caso: "pari" e "dispari" qui si comportano come "positivo" e "negativo" rispetto al prodotto, un parallelo che vale la pena notare.

### Ogni funzione si scompone in una parte pari e una dispari

Un fatto sorprendente: **ogni** funzione $f$ (con dominio simmetrico) si può scrivere come somma di una funzione pari e una dispari:

$$f(x) = \underbrace{\frac{f(x)+f(-x)}{2}}_{\text{parte pari}} + \underbrace{\frac{f(x)-f(-x)}{2}}_{\text{parte dispari}}$$

Verifichiamo che il primo addendo sia davvero pari: chiamandolo $p(x)=\frac{f(x)+f(-x)}{2}$, calcoliamo $p(-x)=\frac{f(-x)+f(x)}{2}=p(x)$ ✓ (abbiamo solo scambiato l'ordine della somma, proprietà commutativa, notebook 0.5).

````python
import sympy as sp

x = sp.symbols('x', real=True)

def è_pari(f):
    return sp.simplify(f.subs(x, -x) - f) == 0

def è_dispari(f):
    return sp.simplify(f.subs(x, -x) + f) == 0

funzioni_test = {
    "x^2": x**2,
    "x^3": x**3,
    "x^2+x": x**2 + x,
    "cos(x)": sp.cos(x),
    "sin(x)": sp.sin(x),
}

for nome, f in funzioni_test.items():
    print(f"{nome}: pari={è_pari(f)}, dispari={è_dispari(f)}")

# Scomposizione in parte pari e parte dispari
f = sp.exp(x)  # funzione né pari né dispari
parte_pari = (f + f.subs(x, -x)) / 2
parte_dispari = (f - f.subs(x, -x)) / 2
print(f"\nf(x)=e^x -> parte pari: {sp.simplify(parte_pari)}, parte dispari: {sp.simplify(parte_dispari)}")
print("Somma delle due parti:", sp.simplify(parte_pari + parte_dispari), " (deve ridare f(x))")
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $f(x)=\dfrac{x^3}{x^2+1}$ è pari, dispari, o nessuna delle due.

**Ragionamento:**

**Passo 1 — calcoliamo $f(-x)$**, sostituendo $-x$ al posto di $x$ in ogni occorrenza:

$$f(-x) = \frac{(-x)^3}{(-x)^2+1}$$

**Passo 2 — semplifichiamo numeratore e denominatore separatamente.** Numeratore: $(-x)^3=-x^3$ (un esponente dispari "conserva" il segno negativo, notebook 0.6). Denominatore: $(-x)^2=x^2$ (un esponente pari lo "cancella"):

$$f(-x) = \frac{-x^3}{x^2+1}$$

**Passo 3 — confrontiamo con $f(x)$ e $-f(x)$.** Notiamo che:

$$f(-x) = \frac{-x^3}{x^2+1} = -\frac{x^3}{x^2+1} = -f(x)$$

**Conclusione**: $f(-x)=-f(x)$ per ogni $x$: la funzione è **dispari**.

**Perché il denominatore "non conta" ai fini della parità qui, ma non è un caso generale:** in questo esempio, il denominatore $x^2+1$ è esso stesso una funzione **pari** (esponente pari + costante), quindi resta invariato passando da $x$ a $-x$. Il numeratore, dispari, determina da solo il comportamento del quoziente (coerente con la regola "dispari/pari = dispari", analoga a quella del prodotto vista nella teoria). Se il denominatore fosse stato una funzione dispari o "mista", la conclusione sarebbe potuta essere diversa: ogni caso va verificato con il calcolo diretto, le regole pratiche sono una scorciatoia utile ma non sostituiscono la definizione quando c'è dubbio.

````python
import sympy as sp

x = sp.symbols('x', real=True)
f = x**3 / (x**2 + 1)

f_meno_x = f.subs(x, -x)
print("f(-x) =", sp.simplify(f_meno_x))
print("-f(x) =", sp.simplify(-f))
print("f(-x) == -f(x)?", sp.simplify(f_meno_x - (-f)) == 0)
````

## Riepilogo finale

- **Pari**: $f(-x)=f(x)$, grafico simmetrico rispetto all'asse $y$ (es. $x^2$, $\cos x$).
- **Dispari**: $f(-x)=-f(x)$, grafico simmetrico rispetto all'origine (es. $x^3$, $\sin x$, $\tan x$).
- La maggior parte delle funzioni non è né pari né dispari: sono categorie speciali, non un'alternativa esaustiva come "pari/dispari" per i numeri interi.
- Regole pratiche: polinomi con soli esponenti pari (o dispari) hanno quella parità; per somme e prodotti valgono regole analoghe a quelle dei segni (notebook 0.2).
- Ogni funzione (con dominio simmetrico) si scompone in una parte pari e una dispari, sommandola alla sua "riflessa" $f(-x)$ opportunamente combinata.
