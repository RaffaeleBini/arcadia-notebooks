# 4.9 — Serie armonica

**Prerequisiti:** 4.7 Serie numeriche, 4.4 Successioni monotone.


## Perché ci interessa

Nel notebook 4.7 abbiamo dimostrato che la convergenza di una serie richiede $a_n\to0$, ma abbiamo avvertito che questa condizione **non è sufficiente**. La **serie armonica**, $\sum\frac1n$, è l'esempio classico che lo dimostra: i suoi termini tendono a zero (lentissimamente, ma tendono), eppure la serie **diverge**. È un risultato sorprendente al primo incontro, e la sua dimostrazione — attribuita a Nicola Oresme, XIV secolo — è un piccolo capolavoro di ragionamento elementare.

## Teoria, passo dopo passo

### La serie armonica

$$\sum_{n=1}^\infty \frac1n = 1+\frac12+\frac13+\frac14+\frac15+\dots$$

I termini $\frac1n\to0$ (notebook 4.2): la condizione necessaria del notebook 4.7 è soddisfatta. Eppure, come vedremo, la serie diverge a $+\infty$.

### La dimostrazione per raggruppamento (Oresme)

L'idea: raggruppiamo i termini in blocchi la cui lunghezza raddoppia ogni volta, e mostriamo che **ogni blocco** contribuisce almeno $\frac12$ alla somma totale.

$$1+\underbrace{\frac12}_{\ge\frac12}+\underbrace{\left(\frac13+\frac14\right)}_{\ge\frac12}+\underbrace{\left(\frac15+\frac16+\frac17+\frac18\right)}_{\ge\frac12}+\dots$$

Verifichiamo ogni blocco: $\frac13+\frac14 > \frac14+\frac14=\frac12$ (abbiamo sostituito $\frac13$, più grande, con $\frac14$, più piccolo: la somma può solo diminuire o restare uguale — notebook 1.8, disuguaglianze). Analogamente, $\frac15+\frac16+\frac17+\frac18 > 4\times\frac18=\frac12$ (quattro termini, ciascuno sostituito dal più piccolo del gruppo, $\frac18$).

In generale, il blocco che va da $\frac1{2^k+1}$ a $\frac1{2^{k+1}}$ contiene $2^k$ termini, ciascuno $\ge\frac1{2^{k+1}}$ (il più piccolo del gruppo): la somma del blocco è $\ge2^k\times\frac1{2^{k+1}}=\frac12$.

### Perché questo dimostra la divergenza

Dopo $k$ blocchi (oltre al primo termine $1$), la somma parziale è **almeno** $1+\frac{k}2$ (ogni blocco contribuisce almeno $\frac12$, teoria). Ma $1+\frac k2\to+\infty$ al crescere di $k$ (una successione aritmetica non limitata, notebook 3.17, 3.11): quindi anche le somme parziali della serie armonica, essendo **sempre maggiori o uguali** a una quantità che diverge, devono anch'esse divergere.

**Il ruolo della monotonia**: le somme parziali $S_N$ della serie armonica sono una successione **crescente** (ogni termine aggiunto, $\frac1{N+1}$, è positivo, notebook 3.17). Per il notebook 4.4, una successione crescente o converge (se limitata) o diverge a $+\infty$ (se non limitata) — **non può oscillare**. Avendo appena mostrato che non è limitata (cresce oltre ogni $1+\frac k2$), concludiamo che diverge proprio a $+\infty$: nessuna terza possibilità.

### La lentezza della divergenza

Un fatto sorprendente: la serie armonica diverge, ma **lentissimamente**. Per superare la somma $10$, servono circa $12367$ termini; per superare $20$, servono circa $2.7\times10^8$ termini. Questa lentezza è il motivo per cui, sperimentando solo numericamente (sommando "abbastanza" termini a mano), si potrebbe erroneamente pensare che la serie converga — un monito sull'importanza delle dimostrazioni rigorose rispetto alla sola evidenza numerica.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

# Verifica dei raggruppamenti: ogni blocco contribuisce almeno 1/2
blocco2 = sp.Rational(1,3) + sp.Rational(1,4)
blocco3 = sp.Rational(1,5) + sp.Rational(1,6) + sp.Rational(1,7) + sp.Rational(1,8)
print(f"1/3+1/4 = {blocco2} = {float(blocco2):.4f}  (>= 1/2? {blocco2 >= sp.Rational(1,2)})")
print(f"1/5+...+1/8 = {blocco3} = {float(blocco3):.4f}  (>= 1/2? {blocco3 >= sp.Rational(1,2)})")

# Verifica della lentissima divergenza: quante somme servono per superare 5?
def somma_armonica_fino_a(soglia):
    S, n = 0, 0
    while S < soglia:
        n += 1
        S += 1/n
    return n, S

n_necessario, S_finale = somma_armonica_fino_a(5)
print(f"\nServono {n_necessario} termini per superare 5 (S={S_finale:.4f})")

# Confronto tra il valore reale della somma parziale e la stima 1 + k/2
import numpy as np
for N in [10, 100, 1000, 10000]:
    S_N = sum(1/k for k in range(1, N+1))
    print(f"N={N}: S_N = {S_N:.4f}")
````

## Esercizio svolto e commentato

**Problema:** usa lo stesso metodo di raggruppamento della serie armonica per stabilire se $\sum_{n=1}^\infty\dfrac1{\sqrt n}$ converge o diverge.

**Ragionamento:**

**Passo 1 — verifichiamo la condizione necessaria** (notebook 4.7): $\frac1{\sqrt n}\to0$ (il denominatore cresce senza limite). La condizione è soddisfatta, ma sappiamo (dalla teoria di questo notebook) che **non basta**.

**Passo 2 — confrontiamo con la serie armonica, termine per termine.** Per $n\ge1$: $\sqrt n\le n$ (perché elevare al quadrato entrambi i membri, entrambi non negativi, dà $n\le n^2$, vero per $n\ge1$, notebook 0.6). Quindi, invertendo (per numeri positivi, l'ordine si inverte, notebook 1.8):

$$\frac1{\sqrt n} \ge \frac1n$$

**Passo 3 — applichiamo un principio di confronto** (analogo, nello spirito, alla tecnica del notebook 4.5 per la limitatezza, ma qui per le somme): se ogni termine di $\sum\frac1{\sqrt n}$ è **maggiore o uguale** al corrispondente termine della serie armonica (che sappiamo divergere), anche le somme parziali di $\sum\frac1{\sqrt n}$ superano quelle della serie armonica:

$$\sum_{n=1}^N\frac1{\sqrt n} \ge \sum_{n=1}^N\frac1n$$

**Passo 4 — concludiamo.** Il lato destro diverge a $+\infty$ (teoria di questo notebook); essendo il lato sinistro sempre maggiore o uguale, anche $\sum\frac1{\sqrt n}$ **diverge**.

**Perché questo "confronto diretto" tra serie è un'anticipazione della tecnica generale che vedremo nel prossimo notebook:** invece di ripetere da zero l'argomento di raggruppamento di Oresme (specifico per $\frac1n$), abbiamo sfruttato una serie **già nota** (la serie armonica) come "pietra di paragone": se una serie ha termini sempre maggiori di una serie che diverge, diverge anch'essa. Questo principio, qui usato in un caso specifico, sarà formalizzato come **criterio del confronto** nel notebook 4.10.

````python
import sympy as sp

n = sp.symbols('n', positive=True)

# Verifica della disuguaglianza 1/sqrt(n) >= 1/n per n>=1
for valore in [1, 2, 5, 10, 100]:
    a = 1 / sp.sqrt(valore)
    b = sp.Rational(1, valore)
    print(f"n={valore}: 1/sqrt(n)={float(a):.4f}, 1/n={float(b):.4f}, 1/sqrt(n)>=1/n? {a >= b}")

# Confronto delle somme parziali
def somma_parziale(termine_func, N):
    return sum(termine_func(k) for k in range(1, N+1))

for N in [10, 100, 1000]:
    S_radice = somma_parziale(lambda k: 1/k**0.5, N)
    S_armonica = somma_parziale(lambda k: 1/k, N)
    print(f"N={N}: sum 1/sqrt(n) = {S_radice:.4f}  >=  sum 1/n = {S_armonica:.4f}? {S_radice >= S_armonica}")
````

## Riepilogo finale

- La serie armonica $\sum\frac1n$ diverge, pur avendo termini che tendono a $0$: dimostra che la condizione $a_n\to0$ (notebook 4.7) non è mai sufficiente per la convergenza.
- La dimostrazione di Oresme raggruppa i termini in blocchi che raddoppiano di lunghezza, ciascuno contribuendo almeno $\frac12$: la somma cresce senza limite.
- Le somme parziali crescenti (notebook 4.4) o convergono (se limitate) o divergono a $+\infty$: non c'è una terza possibilità di comportamento oscillante.
- Il confronto termine a termine con una serie di comportamento noto (qui, la serie armonica) permette di dedurre la divergenza (o convergenza) di altre serie, un'anticipazione del criterio del confronto.
