# 4.2 — Limiti di successioni

**Prerequisiti:** 4.1 Successioni (approfondimento), 0.8 Valore assoluto, 0.14 Intervalli.


## Perché ci interessa

Nel notebook 3.17 abbiamo calcolato $a_n=\frac1{n+1}$ per i primi valori di $n$: $1,\ 0.5,\ 0.333\dots,\ 0.25,\dots$ — sempre più vicino a $0$, senza mai raggiungerlo esattamente. Nel notebook 1.10 abbiamo già usato informalmente l'idea di "avvicinarsi indefinitamente" per costruire le potenze con esponente irrazionale. È arrivato il momento di dare a questa idea un nome preciso: il **limite**. È il concetto fondativo di tutta l'analisi matematica, e da qui in avanti lo useremo costantemente.

## Teoria, passo dopo passo

### L'idea intuitiva

Diciamo che la successione $(a_n)$ **tende a $L$** (o **converge a $L$**) se, man mano che $n$ cresce, i termini $a_n$ si avvicinano a $L$ quanto vogliamo, restandoci vicino per sempre da un certo punto in poi. Scriviamo:

$$\lim_{n\to\infty} a_n = L \qquad \text{oppure} \qquad a_n \to L$$

Per $a_n=\frac1{n+1}$: $\lim_{n\to\infty}\frac1{n+1}=0$. Attenzione: nessun termine della successione **è mai** esattamente $0$ (per nessun $n$ naturale, $\frac1{n+1}=0$); il limite descrive **verso dove tende** la successione, non un valore raggiunto.

### La definizione precisa (un primo assaggio)

Cosa significa esattamente "vicino quanto vogliamo, da un certo punto in poi"? Fissiamo una tolleranza $\varepsilon>0$ (piccola quanto vogliamo, la lettera greca epsilon è la scelta standard). Diciamo che $a_n\to L$ se, per **ogni** $\varepsilon>0$, esiste un indice $N$ (che dipende da $\varepsilon$: tolleranze più strette richiederanno indici più grandi) tale che:

$$|a_n-L| < \varepsilon \qquad \text{per ogni } n>N$$

Usando i quantificatori (notebook 0.17): $\forall\varepsilon>0,\ \exists N,\ \forall n>N,\ |a_n-L|<\varepsilon$. Il valore assoluto (notebook 0.8) misura la **distanza** tra $a_n$ e $L$: la condizione dice che questa distanza, da un certo indice $N$ in poi, resta sempre sotto la soglia $\varepsilon$ scelta. Non approfondiamo qui tutte le sottigliezze di questa definizione (ε-δ rigorosa, con le sue tecniche dimostrative): la riprenderemo con tutto il rigore necessario nel Modulo 10. Per ora, l'intuizione geometrica basta: **qualunque intervallo** $(L-\varepsilon,L+\varepsilon)$ (notebook 0.14) attorno a $L$, per quanto piccolo, contiene **tutti** i termini della successione da un certo punto in poi.

### Verifica dell'esempio guida

Per $a_n=\frac1{n+1}\to0$: fissiamo $\varepsilon=0.01$. Vogliamo $\left|\frac1{n+1}-0\right|<0.01$, cioè $\frac1{n+1}<0.01$ (essendo sempre positivo), cioè $n+1>100$, cioè $n>99$. Quindi $N=99$ funziona: da $n=100$ in poi, tutti i termini distano meno di $0.01$ da $0$. Se scegliessimo $\varepsilon$ più piccolo (per esempio $0.0001$), servirebbe un $N$ più grande ($n>9999$) — ma **esiste sempre** un $N$ adeguato, qualunque sia $\varepsilon$: questo è precisamente ciò che rende vero il limite.

### Successioni che non convergono

Non tutte le successioni hanno limite. Due comportamenti tipici:

- **Divergenza a $+\infty$** (o $-\infty$): $a_n=n$ cresce senza limite, non si avvicina a nessun numero finito. Si scrive $\lim_{n\to\infty}n=+\infty$ (un simbolo, non un numero reale, notebook 0.14).
- **Comportamento oscillante** (nessun limite, né finito né infinito): $a_n=(-1)^n$ vale alternativamente $1,-1,1,-1,\dots$, senza mai stabilizzarsi vicino a un valore. Approfondiremo questo caso nel notebook 4.6, con le sottosuccessioni.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

n = sp.symbols('n', integer=True, positive=True)

# Verifica dell'esempio guida: per quale N vale |a_n - 0| < epsilon?
def trova_N(epsilon):
    # 1/(n+1) < epsilon  =>  n+1 > 1/epsilon  =>  n > 1/epsilon - 1
    return sp.ceiling(1/epsilon - 1)

for eps in [0.1, 0.01, 0.0001]:
    N = trova_N(eps)
    print(f"epsilon={eps}: N={N}  (da n={N+1} in poi, la distanza da 0 è sempre < {eps})")

# Verifica numerica diretta
a_n = lambda k: 1/(k+1)
print("\na_100 =", a_n(100), " (deve essere < 0.01)")
print("a_99 =", a_n(99), " (deve essere >= 0.01, il limite N=99 è il più piccolo che funziona)")

# Grafico: la successione si avvicina a 0, rimanendo dentro la "fascia" epsilon
indici = np.arange(1, 60)
valori = 1 / (indici + 1)
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(indici, valori, 'o', markersize=4)
ax.axhline(y=0, color='black', linewidth=0.5)
eps = 0.05
ax.axhspan(-eps, eps, color='green', alpha=0.15, label=f'fascia (-ε,ε), ε={eps}')
ax.legend()
ax.set_title("a_n = 1/(n+1): da un certo N in poi, resta nella fascia ε")
ax.grid(True, alpha=0.3)
plt.show()
````

## Esercizio svolto e commentato

**Problema:** verifica, con la definizione, che $\lim_{n\to\infty}\dfrac{2n+1}{n}=2$, trovando $N$ in funzione di un $\varepsilon$ generico.

**Ragionamento:**

**Passo 1 — impostiamo la condizione della definizione.** Vogliamo $\left|\dfrac{2n+1}{n}-2\right|<\varepsilon$.

**Passo 2 — semplifichiamo l'espressione dentro il valore assoluto:**

$$\frac{2n+1}{n}-2 = \frac{2n+1}{n}-\frac{2n}{n} = \frac{2n+1-2n}{n} = \frac1n$$

(sottrazione di frazioni con denominatore comune, notebook 0.3). La condizione diventa $\left|\dfrac1n\right|<\varepsilon$, cioè $\dfrac1n<\varepsilon$ (essendo $n$ positivo, il valore assoluto non cambia nulla, notebook 0.8).

**Passo 3 — risolviamo per $n$** (disequazione, notebook 1.8, moltiplicando entrambi i membri per $n$ e poi dividendo per $\varepsilon$, entrambi positivi, verso invariato):

$$n > \frac1\varepsilon$$

**Passo 4 — concludiamo.** Basta scegliere $N=\dfrac1\varepsilon$ (o un qualunque numero naturale più grande, come $\lceil\frac1\varepsilon\rceil$, l'arrotondamento per eccesso): per ogni $n>N$, la condizione $\left|\frac{2n+1}n-2\right|<\varepsilon$ è automaticamente soddisfatta.

**Verifica numerica** con $\varepsilon=0.001$: $N=1000$. Per $n=1001$: $\left|\frac{2003}{1001}-2\right|=\left|\frac{2003-2002}{1001}\right|=\frac1{1001}\approx0.000999<0.001$. ✓

**Perché aver trovato una formula per $N$ **in funzione di $\varepsilon$** (non solo per un $\varepsilon$ fissato) è ciò che rende la dimostrazione completa:** la definizione richiede che **per ogni** $\varepsilon>0$ (notebook 0.17, quantificatore universale) esista un $N$ adeguato. Aver mostrato una formula esplicita $N=\frac1\varepsilon$ garantisce che, qualunque tolleranza venga richiesta, sappiamo sempre costruire l'indice giusto — non abbiamo verificato solo un caso particolare, ma l'intera famiglia infinita di casi possibili.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)
espressione = (2*n + 1) / n - 2
semplificata = sp.simplify(espressione)
print("Espressione semplificata:", semplificata)

# Verifica numerica per un epsilon specifico
epsilon = sp.Rational(1, 1000)
N = 1 / epsilon
print(f"\nCon epsilon={float(epsilon)}, N={N}")

n_test = N + 1
valore = (2*n_test + 1) / n_test
distanza = abs(valore - 2)
print(f"Per n={n_test}: |a_n - 2| = {distanza} < {float(epsilon)}? {distanza < epsilon}")
````

## Riepilogo finale

- $a_n\to L$ significa che i termini della successione si avvicolano a $L$ quanto vogliamo, restandoci vicino definitivamente: per ogni $\varepsilon>0$ esiste $N$ tale che $|a_n-L|<\varepsilon$ per ogni $n>N$.
- Trovare $N$ **in funzione di $\varepsilon$** (non solo per un valore fissato) è ciò che dimostra rigorosamente un limite.
- Non tutte le successioni convergono: possono divergere a $\pm\infty$, oppure oscillare senza stabilizzarsi.
- Questo concetto, qui introdotto in modo intuitivo, sarà formalizzato con tutto il rigore nel Modulo 10 (definizione ε-δ completa).
