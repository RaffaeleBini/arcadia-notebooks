# 4.4 — Successioni monotone

**Prerequisiti:** 4.3 Convergenza, 3.9 Funzioni monotone.


## Perché ci interessa

Nel notebook 4.3 abbiamo visto che convergenza implica limitatezza, ma non viceversa. Ci manca uno strumento per **garantire** la convergenza senza dover indovinare il valore del limite in anticipo. La risposta sta nella monotonia (notebook 3.9): se una successione è monotona **e** limitata, converge **sempre** — un teorema che userai moltissimo, perché a volte è più facile verificare monotonia e limitatezza che calcolare esplicitamente un limite.

## Teoria, passo dopo passo

### Il teorema delle successioni monotone

> Se $(a_n)$ è **crescente** (notebook 3.9: $a_n\le a_{n+1}$ per ogni $n$) e **limitata superiormente** (notebook 3.10), allora $(a_n)$ **converge**.

Analogamente, se $(a_n)$ è decrescente e limitata inferiormente, converge. Questo risultato — che ci risparmia di dover "indovinare" il limite — dipende in modo essenziale dalla **completezza** dei numeri reali (la proprietà che non abbiamo "buchi" sulla retta, notebook 0.4, 0.18): la dimostrazione rigorosa richiede lo strumento dell'estremo superiore, che formalizzeremo nel Modulo 5. Per ora ne cogliamo l'intuizione.

### Intuizione geometrica

Immagina una successione crescente: $a_0\le a_1\le a_2\le\dots$, che "sale" sempre (o resta ferma), senza mai scendere. Se questa salita è **limitata** da un tetto $M$ (nessun termine supera $M$), i termini non possono "sfuggire verso l'infinito": devono accumularsi da qualche parte, avvicinandosi sempre di più a un valore limite — che sarà, precisamente, il più piccolo tetto possibile (l'estremo superiore dell'immagine, un'anticipazione del notebook 3.10 e del Modulo 5).

### Perché questo teorema è così utile

A differenza della definizione ε-N (notebook 4.2), che richiede di **conoscere già** il candidato limite $L$ per verificarlo, il teorema delle successioni monotone garantisce l'**esistenza** del limite senza doverlo calcolare esplicitamente. Questo è prezioso per successioni definite ricorsivamente (notebook 4.1), dove trovare una formula esplicita per il limite può essere difficile o impossibile, ma verificare monotonia e limitatezza è spesso alla portata.

### Esempio guida: una successione ricorsiva

Consideriamo $a_1=1$, $a_{n+1}=\sqrt{2+a_n}$ (una ricorsione "innestata", diversa da quelle additive/moltiplicative del notebook 4.1). Calcoliamo i primi termini: $a_1=1$, $a_2=\sqrt3\approx1.73$, $a_3=\sqrt{3.73}\approx1.93$, $a_4\approx1.98$: sembra crescente e limitata (si avvicina a $2$).

**Verifichiamo la limitatezza** (per induzione, notebook 0.20): se $a_n<2$, allora $a_{n+1}=\sqrt{2+a_n}<\sqrt{2+2}=2$. Con $a_1=1<2$ come base, per induzione $a_n<2$ per ogni $n$: la successione è limitata superiormente da $2$.

**Verifichiamo la monotonia**: se $a_n<2$, allora $2+a_n<4$, quindi $a_{n+1}=\sqrt{2+a_n}<2$; inoltre si può mostrare che $a_{n+1}>a_n$ (equivalente a $2+a_n>a_n^2$, cioè $a_n^2-a_n-2<0$, cioè $(a_n-2)(a_n+1)<0$, vero perché $a_n<2$ e $a_n>-1$): la successione è crescente.

Per il teorema, $(a_n)$ **converge**. Chiamando $L$ il limite, e usando che $a_{n+1}\to L$ tanto quanto $a_n\to L$ (la "coda spostata" ha lo stesso limite), passiamo al limite nella relazione ricorsiva: $L=\sqrt{2+L}$, cioè $L^2=2+L$, cioè $L^2-L-2=0$, che scomposta dà $(L-2)(L+1)=0$: $L=2$ oppure $L=-1$. Essendo $a_n>0$ sempre, scartiamo $L=-1$: $L=2$.

````python
import sympy as sp

# Generiamo numericamente la successione ricorsiva
def successione_ricorsiva(N):
    a = [1]
    for _ in range(N):
        a.append(sp.sqrt(2 + a[-1]))
    return a

termini = successione_ricorsiva(8)
for i, t in enumerate(termini, start=1):
    print(f"a_{i} = {t} ≈ {float(t):.6f}")

# Verifica che il limite candidato L=2 soddisfi L = sqrt(2+L)
L = sp.symbols('L', positive=True)
soluzioni = sp.solve(sp.Eq(L, sp.sqrt(2 + L)), L)
print("\nSoluzioni di L = sqrt(2+L):", soluzioni)

# Verifica di monotonia e limitatezza sui valori calcolati
valori_numerici = [float(t) for t in termini]
crescente = all(valori_numerici[i] <= valori_numerici[i+1] for i in range(len(valori_numerici)-1))
limitata = all(v < 2 for v in valori_numerici)
print(f"\nCrescente sui termini calcolati? {crescente}")
print(f"Limitata da 2 sui termini calcolati? {limitata}")
````

## Esercizio svolto e commentato

**Problema:** studia la successione $a_n=\left(1+\dfrac1n\right)^n$ (per $n\ge1$): calcola i primi termini e discuti perché il teorema delle successioni monotone garantisce che converga (senza calcolare esplicitamente il limite).

**Ragionamento:**

**Passo 1 — calcoliamo alcuni termini:**

$$a_1=(1+1)^1=2 \qquad a_2=(1.5)^2=2.25 \qquad a_{10}\approx2.594 \qquad a_{100}\approx2.705 \qquad a_{1000}\approx2.7169$$

Sembra crescente, e i valori si avvicinano a un numero specifico.

**Passo 2 — accettiamo (senza dimostrarlo qui, richiederebbe strumenti che avremo nel Modulo 5) che questa successione è crescente e limitata superiormente** (per esempio da $3$: si può mostrare con la formula del binomio, che non trattiamo in questo corso, che $a_n<3$ per ogni $n$).

**Passo 3 — applichiamo il teorema delle successioni monotone.** Essendo crescente e limitata superiormente, per il teorema **converge**: esiste un limite finito $L$, anche se non l'abbiamo ancora calcolato esplicitamente.

**Passo 4 — riconosciamo il limite.** Questo limite è **per definizione** il numero $e$ (notebook 1.11!): $e=\lim_{n\to\infty}\left(1+\frac1n\right)^n\approx2.71828$. Questa è la costruzione rigorosa del numero che, nel notebook 1.11, avevamo introdotto solo come "un numero speciale $\approx2.71828$": ora ne vediamo l'origine precisa, come limite di una successione monotona e limitata.

**Perché il teorema delle successioni monotone è indispensabile qui, e non un dettaglio tecnico:** senza questo teorema, non avremmo modo di **sapere** che il limite esiste prima di calcolarlo — e in questo caso, calcolare direttamente il valore numerico del limite (senza sapere già che esiste) sarebbe circolare o comunque molto più delicato. Il teorema ci garantisce l'esistenza "gratuitamente", sulla sola base di monotonia e limitatezza, permettendoci poi di *definire* $e$ come quel limite, con la certezza che tale limite esista davvero.

````python
import sympy as sp

n = sp.symbols('n', positive=True)
a_n = (1 + 1/n)**n

for valore in [1, 2, 10, 100, 1000, 10000]:
    print(f"a_{valore} = {float(a_n.subs(n, valore)):.6f}")

print(f"\ne (valore di riferimento) = {float(sp.E):.6f}")

# Verifica con SymPy che il limite sia esattamente e
limite = sp.limit(a_n, n, sp.oo)
print("\nLimite calcolato da SymPy:", limite)
````

## Riepilogo finale

- **Teorema delle successioni monotone**: una successione monotona (crescente o decrescente) e limitata converge sempre — senza bisogno di conoscere in anticipo il valore del limite.
- Questo teorema si basa sulla completezza dei numeri reali (notebook 0.4, 0.18), formalizzata pienamente solo nel Modulo 5.
- Per successioni ricorsive, si può spesso dimostrare monotonia e limitatezza per induzione (notebook 0.20), poi trovare il valore del limite "passando al limite" nella relazione ricorsiva stessa.
- Il numero $e$ (notebook 1.11) si definisce rigorosamente come il limite della successione monotona e limitata $\left(1+\frac1n\right)^n$.
