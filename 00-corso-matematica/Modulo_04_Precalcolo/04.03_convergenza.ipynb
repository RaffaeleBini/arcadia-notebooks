## Perché ci interessa

Nel notebook 4.2 abbiamo definito cosa significa $a_n\to L$, ed esaminato un esempio caso per caso. Ma per usare i limiti in modo efficiente, ci servono **proprietà generali**: il limite è unico? Cosa succede se sommiamo due successioni convergenti? Una successione convergente deve essere limitata (notebook 3.10)? Rispondere a queste domande ci dà gli "strumenti di calcolo" per i limiti, evitando di dover tornare alla definizione ε-N ogni volta.

## Teoria, passo dopo passo

### Unicità del limite

Una successione convergente ha **un solo** limite: se $a_n\to L_1$ e $a_n\to L_2$, allora $L_1=L_2$. Questo giustifica la notazione $\lim_{n\to\infty}a_n$ (con l'articolo determinativo implicito "il" limite): non ha senso parlare de "un" limite se ce ne fosse più di uno.

**Intuizione della dimostrazione** (per assurdo, notebook 0.20): se $L_1\ne L_2$, la loro distanza $|L_1-L_2|$ è un numero positivo fissato. Ma se $a_n$ si avvicina sia a $L_1$ sia a $L_2$ quanto vogliamo, da un certo punto in poi dovrebbe stare vicinissimo a **entrambi** contemporaneamente — impossibile se $L_1$ e $L_2$ sono punti diversi e la vicinanza richiesta è più piccola della metà della loro distanza. Una contraddizione, che dimostra $L_1=L_2$.

### Convergenza implica limitatezza

Se $(a_n)$ converge, allora $(a_n)$ è **limitata** (notebook 3.10). Intuizione: definitivamente (da un certo $N$ in poi) i termini stanno vicinissimi a $L$, quindi limitati in quella "coda"; i primi $N$ termini sono comunque un numero **finito** di valori, sempre limitati banalmente. L'unione di due insiemi limitati è limitata.

**Attenzione**: l'implicazione va in un solo verso! Una successione limitata **non** deve necessariamente convergere: $a_n=(-1)^n$ è limitata (sempre tra $-1$ e $1$) ma non converge (notebook 4.2, comportamento oscillante).

### Algebra dei limiti

Se $a_n\to L$ e $b_n\to M$ (entrambe convergenti), allora:

$$a_n+b_n \to L+M \qquad a_n\cdot b_n\to L\cdot M \qquad \frac{a_n}{b_n}\to\frac{L}{M} \ (\text{se } M\ne0)$$

Queste proprietà (che non dimostriamo qui in dettaglio, ma useremo costantemente) permettono di calcolare limiti complicati **scomponendoli** in pezzi più semplici, invece di tornare sempre alla definizione ε-N. Esempio: $\lim_{n\to\infty}\frac{2n+1}{n}$ (notebook 4.2, verificato con la definizione) si può ora calcolare più rapidamente: $\frac{2n+1}{n}=2+\frac1n$, e $\frac1n\to0$ (notebook 4.2), quindi per l'algebra dei limiti (somma): $2+\frac1n\to2+0=2$ — stesso risultato, ottenuto senza rifare tutto il lavoro ε-N.

### Forme indeterminate: un'avvertenza importante

L'algebra dei limiti **non funziona sempre direttamente**. Se $a_n\to+\infty$ e $b_n\to+\infty$, cosa possiamo dire di $a_n-b_n$? Non basta sapere che entrambe divergono: il risultato dipende da **come** divergono. Esempio: $a_n=n+5$, $b_n=n$: $a_n-b_n=5\to5$. Ma $a_n=2n$, $b_n=n$: $a_n-b_n=n\to+\infty$. Stesso schema "$\infty-\infty$", risultati completamente diversi: si chiama **forma indeterminata**, e richiede sempre un'analisi caso per caso (tipicamente, semplificando l'espressione prima di calcolare il limite, come nell'esempio sopra).

````python
import sympy as sp

n = sp.symbols('n', positive=True)

# Verifica dell'algebra dei limiti con SymPy
limite_a = sp.limit(1/n, n, sp.oo)
limite_b = sp.limit(2 + 1/n, n, sp.oo)
print(f"lim (1/n) = {limite_a}")
print(f"lim (2 + 1/n) = {limite_b}  (coerente con 2 + lim(1/n) = 2+0 = 2)")

# Esempio di forma indeterminata infinito meno infinito, con risultati diversi
a1, b1 = n + 5, n
a2, b2 = 2*n, n
print(f"\nlim (a1-b1) con a1=n+5, b1=n: {sp.limit(a1-b1, n, sp.oo)}")
print(f"lim (a2-b2) con a2=2n, b2=n: {sp.limit(a2-b2, n, sp.oo)}")
print("Stesso schema 'infinito - infinito', risultati diversi: è una forma indeterminata")

# Verifica: successione limitata ma non convergente
import numpy as np
a_oscillante = [(-1)**k for k in range(10)]
print("\n(-1)^n, primi termini:", a_oscillante, "-> limitata ma non convergente")
````

## Esercizio svolto e commentato

**Problema:** calcola $\lim_{n\to\infty}\dfrac{3n^2+2n}{n^2-1}$, usando l'algebra dei limiti (non la definizione ε-N).

**Ragionamento:**

**Passo 1 — riconosciamo la forma indeterminata.** Sia numeratore sia denominatore tendono a $+\infty$ (sono polinomi di grado 2 con coefficiente principale positivo, notebook 3.11): siamo di fronte a una forma "$\frac\infty\infty$", un'altra forma indeterminata (analoga a "$\infty-\infty$" della teoria) che richiede un'analisi caso per caso.

**Passo 2 — la strategia standard: raccogliamo la potenza di $n$ di grado più alto**, sia al numeratore sia al denominatore (notebook 1.3, raccoglimento a fattor comune):

$$\frac{3n^2+2n}{n^2-1} = \frac{n^2\left(3+\frac2n\right)}{n^2\left(1-\frac1{n^2}\right)}$$

**Passo 3 — semplifichiamo il fattore comune $n^2$** (lecito, $n^2\ne0$ per $n\ge1$, notebook 1.4):

$$= \frac{3+\frac2n}{1-\frac1{n^2}}$$

**Passo 4 — applichiamo l'algebra dei limiti.** Sappiamo che $\frac2n\to0$ e $\frac1{n^2}\to0$ (notebook 4.2, esercizio 4). Quindi:

$$\lim_{n\to\infty}\frac{3+\frac2n}{1-\frac1{n^2}} = \frac{3+0}{1-0} = 3$$

(qui il quoziente dei limiti è lecito, perché il limite del denominatore, $1$, è diverso da $0$ — la condizione richiesta nell'algebra dei limiti).

**Perché "raccogliere la potenza più alta" trasforma sempre una forma indeterminata $\frac\infty\infty$ (tra polinomi) in un calcolo diretto:** dopo aver raccolto e semplificato, tutti i termini "pericolosi" (quelli che tendevano a infinito) sono stati trasformati in termini del tipo $\frac{c}{n^k}\to0$ (notebook 4.2): l'espressione risultante è un rapporto di limiti **finiti e noti**, calcolabile direttamente con l'algebra dei limiti, senza più alcuna indeterminazione.

````python
import sympy as sp

n = sp.symbols('n', positive=True)
espressione = (3*n**2 + 2*n) / (n**2 - 1)
limite = sp.limit(espressione, n, sp.oo)
print("Limite:", limite)
````

## Riepilogo finale

- Il limite di una successione, se esiste, è **unico**.
- Convergenza implica limitatezza (notebook 3.10), ma non viceversa: esistono successioni limitate che non convergono (es. $(-1)^n$).
- L'**algebra dei limiti** (somma, prodotto, quoziente) permette di calcolare limiti scomponendo espressioni complesse in pezzi più semplici e noti.
- Le **forme indeterminate** (come $\infty-\infty$ o $\frac\infty\infty$) richiedono sempre una manipolazione algebrica preliminare (raccoglimento, razionalizzazione) prima di poter applicare l'algebra dei limiti.
