# 4.7 — Serie numeriche

**Prerequisiti:** 4.3 Convergenza, 4.4 Successioni monotone.


## Perché ci interessa

Ha senso sommare **infiniti** numeri? A prima vista sembra assurdo: $1+1+1+\dots$ cresce senza limite. Ma $\frac12+\frac14+\frac18+\dots$? Sommando sempre più termini, il totale si avvicina a $1$ senza mai superarlo (lo verificheremo). Le **serie numeriche** danno un significato preciso a "somme infinite", usando esattamente lo strumento che abbiamo appena costruito: il limite di una successione.

## Teoria, passo dopo passo

### L'idea: sommare un po' alla volta

Data una successione $(a_n)_{n\ge0}$ (i "termini" della serie), costruiamo una nuova successione, quella delle **somme parziali**:

$$S_0=a_0,\quad S_1=a_0+a_1,\quad S_2=a_0+a_1+a_2,\quad \dots \quad S_N=\sum_{n=0}^{N}a_n$$

Ogni $S_N$ è una somma **finita** (sempre calcolabile, nessun problema): il notebook 4.1 ci ha già mostrato che $(S_N)$ è essa stessa una successione, definita ricorsivamente da $S_0=a_0$, $S_{N+1}=S_N+a_{N+1}$ (una ricorsione additiva).

### Definizione di serie convergente

La **serie** $\sum_{n=0}^\infty a_n$ **converge** alla somma $S$ se la successione delle somme parziali converge (notebook 4.2) a $S$:

$$\sum_{n=0}^\infty a_n = S \quad\Longleftrightarrow\quad \lim_{N\to\infty} S_N = S$$

Se il limite non esiste (o è infinito), la serie **diverge**. Questa definizione riduce completamente lo studio delle serie a quello, già fatto, delle successioni: una serie non è altro che il limite di una particolare successione (quella delle somme parziali), costruita sommando via via i termini di un'altra successione.

### Condizione necessaria (ma non sufficiente): i termini devono tendere a zero

Se $\sum a_n$ converge, allora necessariamente $a_n\to0$. Perché: $a_n=S_n-S_{n-1}$ (la differenza tra somme parziali consecutive), e se entrambe $S_n$ e $S_{n-1}$ tendono allo stesso limite $S$ (essendo $(S_{n-1})$ una sottosuccessione "spostata" di $(S_n)$, stesso limite, notebook 4.6), la loro differenza tende a $S-S=0$ (algebra dei limiti, notebook 4.3).

**Attenzione, questa condizione non è sufficiente**: esistono serie con $a_n\to0$ che comunque divergono. Il caso più famoso — la serie armonica — lo vedremo nel notebook 4.9. Per ora, questa condizione ci serve soprattutto **al contrario**: se $a_n$ **non** tende a $0$, la serie **sicuramente diverge** (contronominale, notebook 0.16) — un test rapido per escludere la convergenza.

### Esempio guida: la serie $\sum\frac1{2^n}$

Calcoliamo le somme parziali: $S_0=1$, $S_1=1+\frac12=1.5$, $S_2=1.75$, $S_3=1.875$, $S_4=1.9375,\dots$ — sembrano avvicinarsi a $2$. Verificheremo con precisione questo valore (e la formula generale) nel prossimo notebook, dedicato proprio alla serie geometrica.

````python
import sympy as sp

n = sp.symbols('n', integer=True, nonnegative=True)

# Costruiamo le somme parziali della serie 1/2^n
def somme_parziali(termine, N):
    parziali = []
    somma = 0
    for k in range(N):
        somma += termine.subs(n, k)
        parziali.append(somma)
    return parziali

termine = sp.Rational(1) / 2**n
parziali = somme_parziali(termine, 10)
print("Somme parziali di sum 1/2^n:")
for k, S in enumerate(parziali):
    print(f"  S_{k} = {S} = {float(S):.6f}")

# Verifica della condizione necessaria: a_n -> 0
limite_termine = sp.limit(termine, n, sp.oo)
print(f"\nlim a_n = {limite_termine}  (condizione necessaria soddisfatta)")

# Calcolo diretto della somma della serie con SymPy
somma_serie = sp.Sum(termine, (n, 0, sp.oo)).doit()
print(f"\nSomma della serie (calcolata da SymPy): {somma_serie}")
````

## Esercizio svolto e commentato

**Problema:** studia la serie $\sum_{n=1}^\infty\left(\dfrac1n-\dfrac1{n+1}\right)$ (una **serie telescopica**), trovando una formula esplicita per $S_N$ e calcolando la somma.

**Ragionamento:**

**Passo 1 — scriviamo esplicitamente la somma parziale**, termine per termine:

$$S_N = \left(1-\frac12\right)+\left(\frac12-\frac13\right)+\left(\frac13-\frac14\right)+\dots+\left(\frac1N-\frac1{N+1}\right)$$

**Passo 2 — osserviamo la struttura "telescopica"**: ogni termine negativo di una parentesi si cancella con il termine positivo della parentesi successiva ($-\frac12$ e $+\frac12$, poi $-\frac13$ e $+\frac13$, e così via, come un cannocchiale che si richiude). Resta solo il **primo** termine positivo e l'**ultimo** termine negativo:

$$S_N = 1-\frac1{N+1}$$

**Passo 3 — verifichiamo su un caso piccolo**, per fiducia nel pattern: $N=3$: $S_3=\left(1-\frac12\right)+\left(\frac12-\frac13\right)+\left(\frac13-\frac14\right)=1-\frac14=\frac34$. Con la formula: $1-\frac1{3+1}=1-\frac14=\frac34$. ✓ Coincide.

**Passo 4 — calcoliamo il limite** (notebook 4.2, un limite già noto):

$$S = \lim_{N\to\infty}\left(1-\frac1{N+1}\right) = 1-0=1$$

La serie converge, e la sua somma è $1$.

**Perché riconoscere una struttura "telescopica" trasforma un problema apparentemente su infiniti termini in un calcolo semplicissimo:** invece di dover sommare infiniti pezzi uno per uno (impossibile direttamente), la cancellazione a catena riduce la somma parziale $S_N$ a **soli due termini** (il primo e l'ultimo), rendendo immediato il calcolo del limite. Questa tecnica — riscrivere il termine generale come una differenza che "si cancella in cascata" — è uno degli strumenti più eleganti per calcolare esplicitamente la somma di alcune serie, quando è applicabile.

````python
import sympy as sp

n, N = sp.symbols('n N', positive=True, integer=True)
termine = 1/n - 1/(n+1)

# Somma simbolica, confermando la struttura telescopica
S_N = sp.Sum(termine, (n, 1, N)).doit()
print("S_N in forma chiusa:", sp.simplify(S_N))

limite = sp.limit(S_N, N, sp.oo)
print("Somma della serie:", limite)
````

## Riepilogo finale

- Una serie $\sum a_n$ converge alla somma $S$ se la successione delle somme parziali $S_N=\sum_{n=0}^N a_n$ converge a $S$ (notebook 4.2): lo studio delle serie si riduce allo studio di una successione.
- Condizione necessaria (non sufficiente): se $\sum a_n$ converge, allora $a_n\to0$; usata al contrario, se $a_n\not\to0$ la serie diverge sicuramente.
- Le **serie telescopiche** (dove il termine generale è una differenza che si cancella a catena) permettono di calcolare $S_N$ in forma chiusa, riducendo il calcolo della somma a un semplice limite.
