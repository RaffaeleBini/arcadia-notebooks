**Prerequisiti:** 4.11 Serie di potenze (introduzione), 4.4 Successioni monotone, 1.11 Esponenziali, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Nel notebook 4.11 abbiamo visto che $\frac1{1-x}$ si può scrivere come serie di potenze. Sorprendentemente, lo stesso vale per funzioni molto diverse: $e^x$, $\sin x$, $\cos x$ — funzioni definite in modo completamente diverso (esponenziale, geometria del cerchio) si possono esprimere come "polinomi infiniti". Questa è un'anteprima di uno dei risultati più belli del Modulo 5: la **formula di Taylor**, che qui vediamo solo nei suoi esempi più importanti, senza ancora la macchina (le derivate) che permette di dimostrarla in generale.

## Teoria, passo dopo passo

### La serie esponenziale

Il numero $e$ è nato, nel notebook 4.4, come limite di $\left(1+\frac1n\right)^n$. Esiste anche un'altra caratterizzazione, tramite una serie di potenze:

$$e^x = \sum_{n=0}^\infty \frac{x^n}{n!} = 1+x+\frac{x^2}{2!}+\frac{x^3}{3!}+\dots$$

Questa serie ha raggio di convergenza $R=\infty$ (calcolato esplicitamente nel notebook 4.11: $\left|\frac{a_{n+1}}{a_n}\right|=\frac1{n+1}\to0$), quindi converge per **ogni** $x$ reale, e la somma coincide **sempre** con $e^x$. Per $x=1$: $e=\sum\frac1{n!}=1+1+\frac12+\frac16+\frac1{24}+\dots$, ritrovando in altra forma il numero $e$ già incontrato (e verificato numericamente nel notebook 4.10, applicando il criterio del rapporto a questa stessa serie).

### Le serie di seno e coseno

Analogamente, si possono scrivere:

$$\sin x = \sum_{n=0}^\infty \frac{(-1)^n x^{2n+1}}{(2n+1)!} = x-\frac{x^3}{3!}+\frac{x^5}{5!}-\dots$$

$$\cos x = \sum_{n=0}^\infty \frac{(-1)^n x^{2n}}{(2n)!} = 1-\frac{x^2}{2!}+\frac{x^4}{4!}-\dots$$

Anche queste convergono per ogni $x$ reale ($R=\infty$). Notiamo che nella serie del seno compaiono **solo** potenze dispari di $x$ (coerente con il fatto che $\sin x$ è una funzione dispari, notebook 2.10, 3.8!), e in quella del coseno **solo** potenze pari (coerente con la sua parità, sempre notebook 2.10, 3.8). Non è affatto un caso: la parità/disparità di una funzione si riflette esattamente sulla presenza di soli esponenti pari/dispari nella sua serie di potenze — la stessa osservazione fatta per i polinomi nel notebook 3.11.

### L'idea di "approssimazione polinomiale"

Fermandoci a un numero finito di termini di ciascuna serie, otteniamo un **polinomio di Taylor**, che approssima la funzione vicino a $x=0$:

$$e^x \approx 1+x+\frac{x^2}2 \qquad (\text{approssimazione di grado 2})$$

$$\sin x \approx x-\frac{x^3}6 \qquad (\text{approssimazione di grado 3})$$

Più termini si aggiungono, più l'approssimazione è precisa (e resta buona su un intervallo più ampio attorno a $0$). Questa idea — approssimare una funzione "complicata" con un polinomio, più semplice da calcolare e manipolare — è al cuore di innumerevoli applicazioni pratiche (calcolo numerico, fisica, ingegneria), ed è ciò che il Modulo 5 formalizzerà con precisione, mostrando **come** trovare i coefficienti di queste serie a partire dalla funzione stessa (tramite le derivate), invece di doverli conoscere già come facciamo qui.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Costruiamo le serie troncate (polinomi di Taylor) per e^x, sin(x), cos(x)
def polinomio_taylor(funzione, ordine):
    return sp.series(funzione, x, 0, ordine).removeO()

for ordine in [3, 5, 7]:
    print(f"e^x (ordine {ordine}):", polinomio_taylor(sp.exp(x), ordine))

print()
for ordine in [4, 6, 8]:
    print(f"sin(x) (ordine {ordine}):", polinomio_taylor(sp.sin(x), ordine))

# Confronto numerico: quanto è buona l'approssimazione?
import numpy as np
x_val = 0.5
approssimazione_e = 1 + x_val + x_val**2/2 + x_val**3/6
valore_esatto_e = np.exp(x_val)
print(f"\ne^0.5: approssimazione (4 termini) = {approssimazione_e:.6f}, valore esatto = {valore_esatto_e:.6f}")

approssimazione_sin = x_val - x_val**3/6
valore_esatto_sin = np.sin(x_val)
print(f"sin(0.5): approssimazione (2 termini) = {approssimazione_sin:.6f}, valore esatto = {valore_esatto_sin:.6f}")
````

## Esercizio svolto e commentato

**Problema:** usa i primi tre termini della serie di $e^x$ per stimare $\sqrt{e}$ (cioè $e^{0.5}$), e confronta con il valore reale.

**Ragionamento:**

**Passo 1 — scriviamo i primi tre termini della serie**, sostituendo $x=0.5$:

$$e^{0.5} \approx 1+0.5+\frac{0.5^2}{2!} = 1+0.5+\frac{0.25}2 = 1+0.5+0.125 = 1.625$$

**Passo 2 — confrontiamo con il valore reale** $\sqrt e\approx1.64872$: l'errore è circa $1.64872-1.625=0.02372$, un errore relativo di circa l'$1.4\%$ — non male per soli tre termini.

**Passo 3 — aggiungiamo un quarto termine** per migliorare la stima:

$$e^{0.5} \approx 1.625+\frac{0.5^3}{3!} = 1.625+\frac{0.125}6 = 1.625+0.0208\dots \approx1.6458$$

Ora l'errore è sceso a circa $1.64872-1.6458\approx0.0029$: un miglioramento di quasi dieci volte, aggiungendo un solo termine.

**Passo 4 — interpretiamo la rapidità di miglioramento.** Ogni termine aggiunto è $\frac{x^n}{n!}$: con $x=0.5$ (piccolo) e $n!$ che cresce rapidissimamente (notebook 4.1, più veloce di ogni esponenziale a base fissa, notebook 3.13), ogni nuovo termine è **molto più piccolo** del precedente — la serie "converge velocemente" per valori di $x$ vicini a $0$.

**Perché l'approssimazione migliora così rapidamente qui, mentre per $x$ grande servirebbero molti più termini:** il termine generico $\frac{x^n}{n!}$ diventa trascurabile quando il fattoriale a denominatore "supera" la potenza $x^n$ al numeratore — e per $x$ piccolo (come $0.5$) questo succede quasi subito (già a $n=3,4$). Per $x$ grande, servirebbero molti più termini prima che $n!$ prenda il sopravvento: la stessa serie, la stessa formula, ma una velocità di convergenza diversa a seconda di **dove** valutiamo la funzione — un'osservazione che tornerà utile nel Modulo 5, quando studieremo con precisione l'errore di queste approssimazioni.

````python
import sympy as sp

x_val = sp.Rational(1, 2)

approssimazione_3 = 1 + x_val + x_val**2/2
approssimazione_4 = approssimazione_3 + x_val**3/6

valore_esatto = sp.sqrt(sp.E)

print(f"Approssimazione a 3 termini: {float(approssimazione_3):.6f}")
print(f"Approssimazione a 4 termini: {float(approssimazione_4):.6f}")
print(f"Valore esatto sqrt(e): {float(valore_esatto):.6f}")

print(f"\nErrore a 3 termini: {float(valore_esatto - approssimazione_3):.6f}")
print(f"Errore a 4 termini: {float(valore_esatto - approssimazione_4):.6f}")
````

## Riepilogo finale

- $e^x$, $\sin x$, $\cos x$ si possono scrivere come serie di potenze con raggio di convergenza infinito: valide per ogni $x$ reale.

- Troncando una serie a un numero finito di termini si ottiene un **polinomio di Taylor**, che approssima la funzione tanto meglio quanti più termini si usano, specialmente vicino al punto di sviluppo ($x=0$ in questi esempi).
