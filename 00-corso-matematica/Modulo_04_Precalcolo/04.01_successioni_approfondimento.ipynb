## Perché ci interessa

Nel notebook 3.17 abbiamo definito le successioni come funzioni $a:\mathbb{N}\to\mathbb{R}$, e visto due famiglie base (aritmetiche e geometriche). Prima di affrontare la domanda centrale di questo modulo — cosa succede a una successione quando $n$ cresce indefinitamente? — vale la pena approfondire alcuni aspetti tecnici: notazioni alternative, successioni definite in modi meno diretti, e un esempio classico (Fibonacci) che mostra quanto le successioni ricorsive possano essere ricche.

## Teoria, passo dopo passo

### Indicizzazione: da dove si parte

Nel notebook 3.17 abbiamo indicizzato le successioni a partire da $n=0$. Molti testi (e molte situazioni pratiche) le indicizzano a partire da $n=1$: $(a_n)_{n\ge1}$. Non c'è una convenzione universale, ma un principio pratico: si sceglie l'indice di partenza che rende la formula più semplice, o che ha senso per il contesto (per esempio, se $a_n$ rappresenta "il numero di oggetti dopo $n$ passi", ha senso partire da $n=0$, il momento iniziale). L'importante è essere coerenti e dichiarare sempre da dove si parte.

### Successioni definite per casi

Come le funzioni reali potevano essere definite a tratti (notebook 1.14, 3.1), anche le successioni possono avere regole diverse a seconda della parità dell'indice o di altre condizioni:

$$a_n = \begin{cases} n^2 & \text{se } n \text{ è pari} \\ -n & \text{se } n \text{ è dispari} \end{cases}$$

Producendo $a_0=0,\ a_1=-1,\ a_2=4,\ a_3=-3,\ a_4=16,\dots$ — una successione che "salta" tra due comportamenti diversi. Vedremo nel notebook 4.6 (sottosuccessioni) come isolare ciascun comportamento separatamente.

### La successione di Fibonacci: ricorsione con due termini precedenti

Nel notebook 3.17, le ricorsioni usavano sempre **un solo** termine precedente ($a_{n+1}$ in funzione di $a_n$). Un esempio storico e importante usa **due** termini precedenti:

$$F_0=0,\quad F_1=1,\quad F_{n}=F_{n-1}+F_{n-2} \quad (n\ge2)$$

Producendo $0,1,1,2,3,5,8,13,21,\dots$: ogni termine è la somma dei due precedenti. Questa successione compare in innumerevoli contesti (crescita di popolazioni, spirali in natura, informatica) ed è un ottimo esempio di come una regola ricorsiva semplicissima possa generare una successione dal comportamento sorprendentemente ricco — per esempio, il rapporto $\frac{F_{n+1}}{F_n}$ si avvicina sempre di più a un valore preciso, il **rapporto aureo** $\varphi=\frac{1+\sqrt5}2$, un fatto che potremo dimostrare rigorosamente solo dopo aver introdotto i limiti (prossimo notebook).

### Formula esplicita da una ricorsione: non sempre immediata

Nel notebook 3.17, passare da ricorsivo a esplicito richiedeva "sommare le differenze" (per ricorsioni additive). Per Fibonacci, la trasformazione è più delicata (richiede risolvere un'equazione di secondo grado sui "tassi di crescita", con tecniche che esulano da questo corso), ma esiste una formula esplicita, la **formula di Binet**:

$$F_n = \frac{\varphi^n-\psi^n}{\sqrt5}, \qquad \varphi=\frac{1+\sqrt5}2,\ \ \psi=\frac{1-\sqrt5}2$$

Sorprendente: una formula con radici e potenze irrazionali produce sempre, per ogni $n$, un numero **intero**! Non dimostriamo questo fatto qui, ma lo verifichiamo numericamente nella cella di codice — un promemoria che una formula "esplicita" può essere assai meno immediata da intuire di una definizione ricorsiva, pur descrivendo esattamente la stessa successione.

````python
import sympy as sp

# Successione di Fibonacci: generazione ricorsiva
def fibonacci_ricorsivo(N):
    F = [0, 1]
    for n in range(2, N):
        F.append(F[-1] + F[-2])
    return F

fib = fibonacci_ricorsivo(15)
print("Primi 15 termini di Fibonacci:", fib)

# Rapporto tra termini successivi: si avvicina al rapporto aureo?
rapporto_aureo = (1 + sp.sqrt(5)) / 2
print(f"\nRapporto aureo (valore esatto): {rapporto_aureo} ≈ {float(rapporto_aureo):.6f}")
for n in [5, 8, 10, 13]:
    rapporto = fib[n+1] / fib[n]
    print(f"F_{n+1}/F_{n} = {fib[n+1]}/{fib[n]} = {rapporto:.6f}")

# Verifica della formula di Binet
n = sp.symbols('n', integer=True, nonnegative=True)
phi = (1 + sp.sqrt(5)) / 2
psi = (1 - sp.sqrt(5)) / 2
formula_binet = (phi**n - psi**n) / sp.sqrt(5)

print("\nVerifica della formula di Binet (deve dare numeri interi):")
for valore in range(8):
    risultato = sp.simplify(formula_binet.subs(n, valore))
    print(f"  F_{valore} dalla formula: {risultato}  (atteso: {fib[valore]})")
````

## Esercizio svolto e commentato

**Problema:** definisci una successione ricorsiva per il fattoriale ($n!=1\times2\times\dots\times n$), e verifica che coincida con la definizione esplicita per i primi termini.

**Ragionamento:**

**Passo 1 — troviamo la relazione ricorsiva.** Il fattoriale di $n$ è "il fattoriale di $n-1$, moltiplicato per $n$": $n!=(n-1)!\times n$. Serve un caso base: $0!=1$ (convenzione standard, coerente col fatto che il prodotto di "nessun fattore" è l'elemento neutro della moltiplicazione, notebook 0.5).

$$a_0=1, \qquad a_n=a_{n-1}\times n \quad (n\ge1)$$

**Passo 2 — calcoliamo i primi termini:**

$$a_0=1$$
$$a_1=a_0\times1=1$$
$$a_2=a_1\times2=2$$
$$a_3=a_2\times3=6$$
$$a_4=a_3\times4=24$$

**Passo 3 — confrontiamo con la definizione esplicita** $a_n=n!=1\times2\times\dots\times n$ (per $n\ge1$): $1!=1$, $2!=2$, $3!=6$, $4!=24$. Coincidono tutti.

**Perché questa ricorsione è **moltiplicativa** e non additiva come quelle viste nel notebook 3.17:** la differenza tra $a_n$ e $a_{n-1}$ non è costante né segue uno schema additivo semplice ($a_1-a_0=0$, ma $a_2-a_1=1$, $a_3-a_2=4$, $a_4-a_3=18$: nessuno schema additivo evidente). Qui la relazione naturale è **moltiplicativa** ($a_n=n\cdot a_{n-1}$), esattamente come per le successioni geometriche (notebook 3.17) — ma con un fattore di moltiplicazione **variabile** ($n$, non costante), che rende il fattoriale una crescita ancora più rapida di qualunque successione geometrica a ragione fissa: lo confermeremo con i limiti nel prossimo notebook, collegandoci a quanto già intuito nel notebook 3.13 sul fattoriale che batte ogni esponenziale.

````python
def fattoriale_ricorsivo(N):
    a = [1]  # a_0 = 1
    for n in range(1, N+1):
        a.append(a[-1] * n)
    return a

valori = fattoriale_ricorsivo(6)
print("Fattoriali (ricorsivo):", valori)

import math
print("Fattoriali (funzione math.factorial):", [math.factorial(n) for n in range(7)])
````

## Riepilogo finale

- L'indicizzazione di una successione (partire da $n=0$ o $n=1$) è una scelta di convenienza, purché dichiarata con chiarezza.
- Le successioni possono essere definite a tratti (come le funzioni, notebook 1.14, 3.1), o con ricorsioni che coinvolgono più di un termine precedente (Fibonacci: due termini).
- Passare da una definizione ricorsiva a una esplicita non è sempre immediato: può richiedere formule sorprendenti (Binet) che nascondono interezza dietro espressioni irrazionali.
- Ricorsioni moltiplicative con fattore variabile (come il fattoriale) crescono più rapidamente di qualunque successione geometrica a ragione fissa.
