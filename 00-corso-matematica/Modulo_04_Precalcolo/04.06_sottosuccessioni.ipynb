# 4.6 — Sottosuccessioni

**Prerequisiti:** 4.2 Limiti di successioni, 4.5 Successioni limitate.


## Perché ci interessa

Nel notebook 4.2 abbiamo visto che $a_n=(-1)^n$ non converge: oscilla tra $1$ e $-1$. Ma se guardiamo **solo** i termini pari ($a_0,a_2,a_4,\dots$, tutti uguali a $1$), questa "selezione" converge banalmente a $1$! L'idea di estrarre alcuni termini da una successione, mantenendo l'ordine, si chiama **sottosuccessione**, ed è uno strumento potente per studiare successioni che non convergono nel loro insieme, ma il cui comportamento diventa chiaro se osservato "a pezzi".

## Teoria, passo dopo passo

### Definizione

Data una successione $(a_n)$, una **sottosuccessione** si ottiene scegliendo un sottoinsieme infinito di indici $n_0<n_1<n_2<\dots$ (strettamente crescente, notebook 3.9) e considerando $(a_{n_k})_{k\in\mathbb{N}}$. In altre parole: si "saltano" alcuni termini, ma **senza mai tornare indietro** nell'ordine originale.

Esempio: per $a_n=(-1)^n$, la scelta $n_k=2k$ (indici pari: $0,2,4,\dots$) dà la sottosuccessione costante $a_{n_k}=1$ per ogni $k$. La scelta $n_k=2k+1$ (indici dispari) dà la sottosuccessione costante $a_{n_k}=-1$.

### Se una successione converge, ogni sua sottosuccessione converge allo stesso limite

> Se $a_n\to L$, allora **ogni** sottosuccessione $(a_{n_k})$ converge anch'essa a $L$.

Intuizione: se i termini di $(a_n)$ si avvicinano a $L$ quanto vogliamo da un certo punto in poi, lo stesso vale a maggior ragione per una selezione di quei termini (che comunque, prendendo $k$ grande, corrisponde a indici $n_k$ altrettanto grandi, essendo $n_k\ge k$ sempre, dato che gli indici sono strettamente crescenti in $\mathbb{N}$).

### Il criterio più utile: due sottosuccessioni con limiti diversi

Questo teorema si usa quasi sempre **al contrario**, per dimostrare che una successione **non converge**:

> Se esistono due sottosuccessioni di $(a_n)$ che convergono a limiti **diversi**, allora $(a_n)$ non converge.

È esattamente lo schema del controesempio (notebook 0.20): se $(a_n)$ convergesse a un qualche $L$, **tutte** le sue sottosuccessioni dovrebbero convergere a quello stesso $L$ (per il risultato precedente) — trovare due sottosuccessioni con limiti diversi è una contraddizione diretta, quindi $(a_n)$ non può convergere. Per $a_n=(-1)^n$: la sottosuccessione degli indici pari converge a $1$, quella degli indici dispari converge a $-1$: **due limiti diversi**, quindi $(-1)^n$ non converge — una dimostrazione rigorosa di un fatto che avevamo solo osservato informalmente nel notebook 4.2.

### Il teorema di Bolzano-Weierstrass (anticipazione)

Un risultato profondo, che enunciamo senza dimostrare (lo vedremo con rigore nel Modulo 10): **ogni successione limitata ammette almeno una sottosuccessione convergente**, anche se la successione originale non converge affatto. Per $a_n=(-1)^n$ (limitata, non convergente): esistono comunque le due sottosuccessioni convergenti trovate sopra. Questo teorema collega la limitatezza (notebook 4.5) a un tipo più debole ma sempre garantito di "buon comportamento", ed è uno degli strumenti fondamentali dell'analisi reale avanzata.

````python
import numpy as np

# Successione oscillante (-1)^n: due sottosuccessioni con limiti diversi
n_vals = np.arange(0, 20)
a_n = (-1.0)**n_vals
print("Successione (-1)^n:", a_n[:10])

sottosuccessione_pari = a_n[0::2]   # indici 0, 2, 4, ...
sottosuccessione_dispari = a_n[1::2]  # indici 1, 3, 5, ...
print("Sottosuccessione indici pari:", sottosuccessione_pari[:5], "-> converge a 1")
print("Sottosuccessione indici dispari:", sottosuccessione_dispari[:5], "-> converge a -1")

# Esempio più articolato: a_n = n se n pari, 1/n se n dispari (comportamento misto)
def a_n_mista(n):
    return n if n % 2 == 0 else 1/n

valori = [a_n_mista(n) for n in range(1, 15)]
print("\nSuccessione mista:", [f"{v:.3f}" for v in valori])
pari = [a_n_mista(n) for n in range(2, 30, 2)]
dispari = [a_n_mista(n) for n in range(1, 30, 2)]
print("Sottosuccessione pari (dovrebbe divergere a +inf):", pari[:5], "...")
print("Sottosuccessione dispari (dovrebbe convergere a 0):", dispari[:5], "...")
````

## Esercizio svolto e commentato

**Problema:** dimostra che $a_n=\sin\left(\dfrac{n\pi}2\right)$ non converge, trovando due sottosuccessioni con limiti diversi.

**Ragionamento:**

**Passo 1 — calcoliamo i primi termini**, usando i valori notevoli del seno (notebook 2.9):

$$a_0=\sin(0)=0,\quad a_1=\sin\left(\frac\pi2\right)=1,\quad a_2=\sin(\pi)=0,\quad a_3=\sin\left(\frac{3\pi}2\right)=-1,\quad a_4=\sin(2\pi)=0,\dots$$

Notiamo uno schema che si ripete: $0,1,0,-1,0,1,0,-1,\dots$ (periodo 4 negli indici, coerente con la periodicità del seno, notebook 2.8, applicata qui a un argomento che cresce a scatti di $\frac\pi2$).

**Passo 2 — costruiamo due sottosuccessioni con comportamento diverso.** Prendiamo gli indici multipli di $4$: $n_k=4k$ (cioè $n=0,4,8,12,\dots$). Allora $a_{4k}=\sin(2k\pi)=0$ per ogni $k$ (essendo $2k\pi$ sempre un multiplo intero di $2\pi$, periodicità del seno): questa sottosuccessione è **costantemente $0$**, quindi converge a $0$.

**Passo 3 — costruiamo una seconda sottosuccessione.** Prendiamo gli indici $n_k=4k+1$ (cioè $n=1,5,9,13,\dots$). Allora $a_{4k+1}=\sin\left(\frac{(4k+1)\pi}2\right)=\sin\left(2k\pi+\frac\pi2\right)=\sin\left(\frac\pi2\right)=1$ (di nuovo per periodicità): questa sottosuccessione è **costantemente $1$**, converge a $1$.

**Conclusione**: abbiamo trovato due sottosuccessioni di $(a_n)$ che convergono a limiti diversi ($0$ e $1$). Per il criterio della teoria, $(a_n)$ **non converge**.

**Perché scegliere sottosuccessioni "passo $4$" (invece di, per esempio, "passo $2$") è la scelta naturale qui:** il periodo dello schema $0,1,0,-1$ è esattamente $4$: scegliendo indici che differiscono di un multiplo di $4$, "congeliamo" la successione sempre nella stessa fase del ciclo, ottenendo sottosuccessioni **costanti** (le più facili possibili da analizzare). Riconoscere il periodo dello schema è sempre il primo passo per scegliere sottosuccessioni utili in casi come questo.

````python
import sympy as sp

n = sp.symbols('n', integer=True, nonnegative=True)
a_n = sp.sin(n * sp.pi / 2)

for valore in range(8):
    print(f"a_{valore} = {a_n.subs(n, valore)}")

k = sp.symbols('k', integer=True, nonnegative=True)
print("\nSottosuccessione n=4k:", sp.simplify(a_n.subs(n, 4*k)))
print("Sottosuccessione n=4k+1:", sp.simplify(a_n.subs(n, 4*k + 1)))
````

## Riepilogo finale

- Una sottosuccessione si ottiene selezionando un sottoinsieme infinito di indici, mantenendo l'ordine originale.
- Se $a_n\to L$, ogni sottosuccessione converge allo stesso $L$; usato al contrario, **due sottosuccessioni con limiti diversi dimostrano che la successione non converge**.
- Il teorema di Bolzano-Weierstrass (anticipato, rigoroso nel Modulo 10): ogni successione limitata ha almeno una sottosuccessione convergente, anche se la successione intera non converge.
- Riconoscere schemi periodici (come per $\sin\frac{n\pi}2$) aiuta a scegliere sottosuccessioni "congelate" su un singolo valore, le più semplici da analizzare.
