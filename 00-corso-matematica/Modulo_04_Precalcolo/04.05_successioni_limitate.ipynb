# 4.5 — Successioni limitate

**Prerequisiti:** 3.10 Funzioni limitate, 4.3 Convergenza, 4.4 Successioni monotone.


## Perché ci interessa

Nei notebook 4.3 e 4.4 la limitatezza è comparsa due volte, con ruoli diversi: come **conseguenza necessaria** della convergenza (4.3: convergente ⟹ limitata) e come **ipotesi** del teorema delle successioni monotone (4.4: monotona + limitata ⟹ convergente). Vale la pena fermarsi e mettere in ordine questi collegamenti, con qualche tecnica pratica in più per **dimostrare** che una successione è limitata — spesso il passo preliminare per applicare il teorema del notebook 4.4.

## Teoria, passo dopo passo

### Riepilogo dei collegamenti logici

Mettiamo in fila le implicazioni viste finora, per chiarire cosa implica cosa (e cosa **non** implica):

$$\text{monotona} + \text{limitata} \ \Rightarrow\ \text{convergente} \ \Rightarrow\ \text{limitata}$$

Da questa catena: **ogni** successione convergente è limitata (prima implicazione a destra), ma **non** ogni successione limitata converge (serve anche la monotonia, o comunque qualche altra condizione — il notebook 4.6 mostrerà come le sottosuccessioni permettono di trattare anche successioni limitate non monotone). E una successione monotona **non limitata** semplicemente diverge a $\pm\infty$ (non oscilla: la monotonia da sola esclude il comportamento oscillante, notebook 4.2).

### Tecniche per dimostrare la limitatezza

**Per induzione** (già usata nel notebook 4.4): si dimostra che $a_1$ soddisfa il limite voluto, poi che se $a_n$ lo soddisfa anche $a_{n+1}$ lo soddisfa — lo schema del notebook 0.20, applicato a una disuguaglianza invece che a un'equazione.

**Per confronto con una successione nota**: se riusciamo a mostrare $|a_n|\le b_n$ per una successione $(b_n)$ già nota essere limitata, allora anche $(a_n)$ è limitata. Esempio: $a_n=\frac{\sin n}{n}$: sappiamo $|\sin n|\le1$ per ogni $n$ (notebook 2.12, immagine del seno), quindi $\left|\frac{\sin n}n\right|\le\frac1n\le1$ per $n\ge1$: limitata.

**Direttamente dalla formula esplicita**: se $a_n$ è un rapporto di polinomi (notebook 3.12) dello stesso grado, sappiamo già (dal notebook 4.3, tecnica del raccoglimento) che converge, quindi (per l'implicazione della teoria) è automaticamente limitata.

### Successioni limitate e valore assoluto

Ricordiamo (notebook 3.10, ora specializzato alle successioni): $(a_n)$ è limitata se esiste $M>0$ tale che $|a_n|\le M$ per **ogni** $n$. Questa singola condizione (con il valore assoluto) racchiude sia la limitatezza superiore sia quella inferiore in una volta sola (notebook 0.8: $|a_n|\le M \iff -M\le a_n\le M$).

````python
import sympy as sp
import numpy as np

n = sp.symbols('n', positive=True, integer=True)

# Tecnica del confronto: |sin(n)/n| <= 1/n
def a_n_num(k):
    return np.sin(k) / k

valori = [a_n_num(k) for k in range(1, 20)]
print("Valori di sin(n)/n:", [f"{v:.4f}" for v in valori[:10]])
print("Tutti con |a_n| <= 1?", all(abs(v) <= 1 for v in valori))
print("Tutti con |a_n| <= 1/n?", all(abs(valori[k-1]) <= 1/k for k in range(1, 20)))

# Verifica per induzione (esempio del notebook 4.4): a_{n+1}=sqrt(2+a_n), limitata da 2
def verifica_limitata_per_induzione(N):
    a = 1  # a_1 = 1 < 2
    for _ in range(N):
        assert a < 2, "la successione ha superato il limite ipotizzato!"
        a = (2 + a) ** 0.5
    return True

print("\nLimitatezza per induzione verificata su 50 passi?", verifica_limitata_per_induzione(50))
````

## Esercizio svolto e commentato

**Problema:** dimostra che $a_n=\dfrac{n}{n^2+1}$ è limitata, usando la tecnica del confronto.

**Ragionamento:**

**Passo 1 — osserviamo che $a_n>0$ per ogni $n\ge1$** (rapporto di due quantità positive): possiamo lavorare direttamente con $a_n$, senza bisogno del valore assoluto.

**Passo 2 — cerchiamo una maggiorazione semplice.** Notiamo che $n^2+1>n^2$ (aggiungendo $1$ a $n^2$, notebook 0.5), quindi, per una frazione con lo stesso numeratore positivo ma denominatore più grande, il valore della frazione è più piccolo (proprietà delle disuguaglianze con frazioni, coerente col notebook 1.9):

$$a_n = \frac{n}{n^2+1} < \frac{n}{n^2} = \frac1n$$

**Passo 3 — usiamo che $\frac1n\le1$ per $n\ge1$** (notebook 4.2): quindi $a_n<\frac1n\le1$.

**Passo 4 — concludiamo.** Avendo $0<a_n<1$ per ogni $n\ge1$, la successione è limitata, con $M=1$ come maggiorante valido (in realtà anche $M=\frac12$ funzionerebbe come vedremo studiando il massimo con le derivate nel Modulo 5, ma non ci serve il maggiorante più stretto possibile: **un qualunque** $M$ che funzioni è sufficiente per dimostrare la limitatezza, teoria del notebook 3.10).

**Perché non serve trovare il maggiorante "migliore" (il più piccolo possibile) per dimostrare la limitatezza:** la definizione di limitatezza (notebook 3.10) richiede solo l'**esistenza** di un qualche $M$ che funzioni, non che sia il più piccolo possibile. Una maggiorazione "grossolana" ma facile da dimostrare (come $a_n<\frac1n\le1$) è spesso preferibile a una più precisa ma più complicata da derivare, perché il nostro obiettivo qui è solo stabilire la limitatezza, non trovarne il valore esatto (l'estremo superiore, che è un problema diverso, trattato nel Modulo 5).

````python
import sympy as sp
from sympy.calculus.util import function_range

n = sp.symbols('n', positive=True)
a_n = n / (n**2 + 1)

# Verifica della maggiorazione a_n < 1/n
differenza = sp.simplify(sp.Rational(1,1)/n - a_n)
print("1/n - a_n =", differenza, " (deve essere sempre positivo per n>0)")

# Immagine esatta (per confronto, con gli strumenti del notebook 3.4)
immagine = function_range(a_n, n, sp.Interval(1, sp.oo))
print("Immagine su n>=1:", immagine)
````

## Riepilogo finale

- Catena di implicazioni: monotona + limitata $\Rightarrow$ convergente $\Rightarrow$ limitata (ma limitata da sola non basta per la convergenza).
- Tecniche per dimostrare la limitatezza: per induzione (notebook 0.20), per confronto con una successione nota (usando il valore assoluto, notebook 0.8), o direttamente dalla struttura della formula (es. rapporti di polinomi, notebook 3.12).
- Per la definizione di limitatezza basta **un qualunque** maggiorante $M$ che funzioni: non serve trovare il più piccolo possibile.
- Una successione non limitata non può mai convergere: è la contronominale della prima implicazione della catena.
