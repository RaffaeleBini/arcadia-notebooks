# 4.11 — Serie di potenze (introduzione)

**Prerequisiti:** 4.10 Criteri di convergenza, 4.8 Serie geometrica, 1.1 Polinomi.


## Perché ci interessa

Finora abbiamo studiato serie di **numeri** fissati: $\sum\frac1{2^n}$, $\sum\frac1n$. Cosa succede se, al posto di un numero, mettiamo una **variabile** $x$? Otteniamo una **serie di potenze**: $\sum a_nx^n$, un "polinomio infinito" (notebook 1.1). Il fatto sorprendente è che, per certi valori di $x$, questa somma infinita converge a un numero preciso — definendo così una **funzione**, un ponte diretto tra le serie e tutte le funzioni studiate nel Modulo 3.

## Teoria, passo dopo passo

### Definizione

Una **serie di potenze** (centrata in $0$) è un'espressione della forma:

$$\sum_{n=0}^\infty a_nx^n = a_0+a_1x+a_2x^2+a_3x^3+\dots$$

dove $(a_n)$ è una successione di coefficienti fissati, e $x$ è una variabile reale. Per ogni valore **specifico** di $x$, questa diventa una serie numerica ordinaria (notebook 4.7), che può convergere o divergere a seconda di $x$.

### L'esempio guida: la serie geometrica come serie di potenze

La serie geometrica del notebook 4.8, $\sum x^n$ (con $a_n=1$ per ogni $n$), è già una serie di potenze! Sappiamo che converge se e solo se $|x|<1$ (la condizione $|q|<1$ del notebook 4.8, ora con $q=x$), e in quel caso:

$$\sum_{n=0}^\infty x^n = \frac1{1-x} \qquad \text{per } |x|<1$$

Questa è la prima "funzione scritta come serie di potenze" che incontriamo: per ogni $x\in(-1,1)$, la somma infinita $1+x+x^2+\dots$ restituisce esattamente il valore $\frac1{1-x}$.

### Il raggio di convergenza

Per una serie di potenze generica, esiste sempre un numero $R\ge0$ (eventualmente $R=\infty$), detto **raggio di convergenza**, tale che:

- La serie converge (assolutamente) per $|x|<R$.
- La serie diverge per $|x|>R$.
- Per $|x|=R$ (i due estremi dell'intervallo), il comportamento va verificato caso per caso: può convergere o divergere.

L'insieme $(-R,R)$ (o $[-R,R]$, o varianti con estremi inclusi/esclusi, notebook 0.14) si chiama **intervallo di convergenza**. Per la serie geometrica, $R=1$: coerente con $|x|<1$.

### Calcolare il raggio con il criterio del rapporto

Applichiamo il criterio del rapporto (notebook 4.10) ai termini $a_nx^n$:

$$\left|\frac{a_{n+1}x^{n+1}}{a_nx^n}\right| = |x|\cdot\left|\frac{a_{n+1}}{a_n}\right|$$

Se $\left|\frac{a_{n+1}}{a_n}\right|\to\ell$ (un limite che dipende solo dai coefficienti, non da $x$), il criterio del rapporto richiede $|x|\cdot\ell<1$ per la convergenza, cioè $|x|<\frac1\ell$: il raggio di convergenza è $R=\frac1\ell$ (con la convenzione che $R=\infty$ se $\ell=0$, e $R=0$ se $\ell=\infty$).

Esempio: per $\sum\frac{x^n}{n!}$ (che vedremo esplicitamente nel prossimo notebook), i coefficienti sono $a_n=\frac1{n!}$, e $\left|\frac{a_{n+1}}{a_n}\right|=\frac1{n+1}\to0$ (calcolo già visto nel notebook 4.10): quindi $R=\infty$, la serie converge per **ogni** $x$ reale.

````python
import sympy as sp

x, n = sp.symbols('x n', real=True)

# Verifica: la serie geometrica come serie di potenze, raggio R=1
a_n = 1  # coefficiente costante
# rapporto |a_(n+1)/a_n| = 1 -> R = 1/1 = 1
print("Serie geometrica: coefficienti costanti, raggio di convergenza R = 1")

# Calcolo del raggio per sum x^n/n! con il criterio del rapporto
a_n_fattoriale = 1 / sp.factorial(n)
rapporto = sp.simplify(a_n_fattoriale.subs(n, n+1) / a_n_fattoriale)
limite_ell = sp.limit(sp.Abs(rapporto), n, sp.oo)
print(f"\nsum x^n/n!: |a_(n+1)/a_n| = {rapporto}, limite = {limite_ell}")
print("Raggio di convergenza: R = infinito (converge per ogni x)")

# Verifica numerica: la somma parziale di sum x^n per x=0.5 si avvicina a 1/(1-x)
x_val = sp.Rational(1, 2)
for N in [5, 10, 20]:
    S_N = sum(x_val**k for k in range(N+1))
    print(f"N={N}: S_N = {float(S_N):.6f}  (atteso 1/(1-0.5) = {float(1/(1-x_val)):.6f})")
````

## Esercizio svolto e commentato

**Problema:** trova il raggio di convergenza di $\sum_{n=1}^\infty\dfrac{x^n}{n}$.

**Ragionamento:**

**Passo 1 — identifichiamo i coefficienti**: $a_n=\dfrac1n$.

**Passo 2 — calcoliamo il rapporto** $\left|\dfrac{a_{n+1}}{a_n}\right|$:

$$\left|\frac{a_{n+1}}{a_n}\right| = \left|\frac{1/(n+1)}{1/n}\right| = \frac{n}{n+1}$$

**Passo 3 — calcoliamo il limite** (tecnica del raccoglimento, notebook 4.3):

$$\ell = \lim_{n\to\infty}\frac{n}{n+1} = 1$$

**Passo 4 — troviamo il raggio**: $R=\dfrac1\ell=\dfrac11=1$.

**Passo 5 — analizziamo i due estremi**, $x=1$ e $x=-1$, dove il criterio del rapporto non decide (teoria):

- Per $x=1$: la serie diventa $\sum\frac1n$, la **serie armonica** (notebook 4.9): **diverge**.
- Per $x=-1$: la serie diventa $\sum\frac{(-1)^n}n$, una serie a segni alterni. Sebbene non abbiamo sviluppato criteri specifici per serie a segni alterni in questo corso introduttivo, si può mostrare che questa **converge** (le somme parziali oscillano, ma con oscillazioni sempre più piccole, avvicinandosi a un limite — un fenomeno collegato al criterio di Leibniz, che approfondirai in corsi più avanzati).

**Conclusione**: l'intervallo di convergenza è $[-1,1)$: chiuso in $-1$ (converge), aperto in $1$ (diverge).

**Perché il comportamento agli estremi richiede sempre un'analisi separata, e non si può dedurre dal solo raggio di convergenza:** il criterio del rapporto (e quello della radice) determinano $R$ guardando il comportamento "asintotico" dei coefficienti, ma esattamente nei punti $x=\pm R$ il criterio produce $L=1$ (il caso "non decisivo", notebook 4.10): serve sempre sostituire il valore specifico e applicare altri strumenti (come il confronto diretto con la serie armonica, fatto qui) per stabilire cosa succede esattamente al bordo.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)
a_n = 1 / n

rapporto = sp.simplify(a_n.subs(n, n+1) / a_n)
ell = sp.limit(rapporto, n, sp.oo)
print("ell =", ell, " -> R =", 1/ell)

# Verifica del comportamento agli estremi con SymPy
print("\nComportamento in x=1 (serie armonica):", "diverge")
print("Comportamento in x=-1 (serie a segni alterni):")
from sympy import Sum, oo
# Verifica numerica delle somme parziali per x=-1
def somma_x_meno_1(N):
    return sum((-1)**k / k for k in range(1, N+1))
for N in [10, 100, 1000]:
    print(f"  N={N}: S_N = {somma_x_meno_1(N):.6f}")
print("  (le somme parziali oscillano ma sembrano stabilizzarsi vicino a -ln(2) ≈ -0.6931)")
````

## Riepilogo finale

- Una serie di potenze $\sum a_nx^n$ converge per $|x|<R$ (raggio di convergenza) e diverge per $|x|>R$; il comportamento in $x=\pm R$ va analizzato caso per caso.
- Il raggio si calcola con il criterio del rapporto (o della radice, notebook 4.10) applicato ai coefficienti: $R=\frac1\ell$, dove $\ell=\lim\left|\frac{a_{n+1}}{a_n}\right|$.
- La serie geometrica $\sum x^n=\frac1{1-x}$ (per $|x|<1$) è l'esempio più semplice: una serie di potenze che, dentro il suo raggio di convergenza, definisce esattamente una funzione razionale nota (notebook 3.12).
- Le serie di potenze sono il ponte diretto tra le serie numeriche e le funzioni: dentro il raggio di convergenza, la somma della serie **è** una funzione di $x$.
