# 4.10 — Criteri di convergenza

**Prerequisiti:** 4.8 Serie geometrica, 4.9 Serie armonica.


## Perché ci interessa

Nel notebook 4.9 abbiamo usato un confronto termine a termine per stabilire la divergenza di $\sum\frac1{\sqrt n}$, senza calcolare esplicitamente le somme parziali. Questa idea — confrontare una serie sconosciuta con una nota — è così utile da meritare una sistematizzazione: i **criteri di convergenza**. In questo notebook raccogliamo i più usati, che ti eviteranno di dover ricorrere alla definizione ε-N (notebook 4.2) o a trucchi ad hoc (come la telescopia, notebook 4.7) ogni volta.

## Teoria, passo dopo passo

### Criterio del confronto

Se $0\le a_n\le b_n$ per ogni $n$ (da un certo punto in poi), allora:

- Se $\sum b_n$ converge, anche $\sum a_n$ converge (una serie "più piccola" di una che converge non può "scappare all'infinito").
- Se $\sum a_n$ diverge, anche $\sum b_n$ diverge (contronominale della stessa idea).

È esattamente la tecnica usata nel notebook 4.9: confrontando $\frac1{\sqrt n}\ge\frac1n$ con la serie armonica (divergente), abbiamo concluso la divergenza. Il criterio richiede di **trovare** una serie di confronto adeguata: spesso si usano serie geometriche (notebook 4.8) o la serie armonica (notebook 4.9) come "pietre di paragone" standard.

### Criterio del rapporto (d'Alembert)

Per serie a termini positivi, calcoliamo:

$$L = \lim_{n\to\infty}\frac{a_{n+1}}{a_n}$$

(il rapporto tra un termine e il precedente, un'idea analoga al coefficiente angolare per le successioni, ma applicata al "tasso di crescita relativo"). Allora:

- Se $L<1$: la serie **converge**.
- Se $L>1$ (o $L=+\infty$): la serie **diverge**.
- Se $L=1$: il criterio **non decide** (serve un altro metodo).

**Intuizione**: se $L<1$, da un certo punto in poi i termini si comportano come una serie geometrica di ragione circa $L<1$ (che sappiamo convergere, notebook 4.8): il rapporto quasi costante tra termini successivi "imita" una progressione geometrica convergente. Questo criterio è particolarmente efficace quando $a_n$ contiene fattoriali o potenze (dove i rapporti si semplificano molto, notebook 4.1).

### Criterio della radice (Cauchy)

Alternativa al criterio del rapporto, spesso utile quando $a_n$ è già una potenza $n$-esima di qualcosa:

$$L = \lim_{n\to\infty}\sqrt[n]{a_n}$$

Stessa conclusione del criterio del rapporto: $L<1$ converge, $L>1$ diverge, $L=1$ non decide.

### Un esempio guida: applichiamo il criterio del rapporto alla serie del fattoriale

Consideriamo $\sum_{n=1}^\infty\frac1{n!}$ (collegata al numero $e$, notebook 1.11 e 4.4 — la vedremo esplicitamente nel notebook 4.12). Calcoliamo:

$$\frac{a_{n+1}}{a_n} = \frac{1/(n+1)!}{1/n!} = \frac{n!}{(n+1)!} = \frac{n!}{(n+1)\cdot n!} = \frac1{n+1}$$

(usando che $(n+1)!=(n+1)\times n!$, la stessa relazione ricorsiva del notebook 4.1). Il limite:

$$L = \lim_{n\to\infty}\frac1{n+1} = 0 < 1$$

Per il criterio del rapporto, la serie **converge**.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)

# Criterio del rapporto applicato a sum 1/n!
a_n = 1 / sp.factorial(n)
rapporto = sp.simplify(a_n.subs(n, n+1) / a_n)
print("a_(n+1)/a_n =", rapporto)
L = sp.limit(rapporto, n, sp.oo)
print("L =", L, " -> converge (L<1)")

# Verifica numerica delle somme parziali
def somma_fattoriale(N):
    return sum(sp.Rational(1, sp.factorial(k)) for k in range(N+1))

for N in [3, 5, 8, 10]:
    print(f"N={N}: S_N = {float(somma_fattoriale(N)):.8f}")
print(f"\ne - 1 = {float(sp.E - 1):.8f}  (la serie sum_{{n=0}}^inf 1/n! converge esattamente a e)")

# Criterio del rapporto applicato alla serie armonica (deve dare L=1, non decisivo)
a_n_arm = 1/n
rapporto_arm = sp.simplify(a_n_arm.subs(n, n+1) / a_n_arm)
L_arm = sp.limit(rapporto_arm, n, sp.oo)
print(f"\nSerie armonica: L = {L_arm}  (il criterio del rapporto NON decide: sappiamo già che diverge, notebook 4.9)")
````

## Esercizio svolto e commentato

**Problema:** studia la convergenza di $\sum_{n=1}^\infty\dfrac{n^2}{2^n}$ usando il criterio del rapporto.

**Ragionamento:**

**Passo 1 — impostiamo il rapporto** $\dfrac{a_{n+1}}{a_n}$ con $a_n=\dfrac{n^2}{2^n}$:

$$\frac{a_{n+1}}{a_n} = \frac{(n+1)^2/2^{n+1}}{n^2/2^n} = \frac{(n+1)^2}{2^{n+1}} \times \frac{2^n}{n^2} = \frac{(n+1)^2}{n^2} \times \frac{2^n}{2^{n+1}}$$

**Passo 2 — semplifichiamo ciascun fattore separatamente.** Il secondo fattore: $\dfrac{2^n}{2^{n+1}}=\dfrac1{2}$ (proprietà delle potenze, notebook 0.6). Il primo fattore: $\left(\dfrac{n+1}n\right)^2$.

$$\frac{a_{n+1}}{a_n} = \frac12\left(\frac{n+1}n\right)^2 = \frac12\left(1+\frac1n\right)^2$$

**Passo 3 — calcoliamo il limite.** Sappiamo che $\frac1n\to0$ (notebook 4.2), quindi $\left(1+\frac1n\right)^2\to(1+0)^2=1$ (algebra dei limiti, notebook 4.3, applicata al quadrato):

$$L = \lim_{n\to\infty}\frac12\left(1+\frac1n\right)^2 = \frac12\times1 = \frac12$$

**Passo 4 — concludiamo.** $L=\frac12<1$: per il criterio del rapporto, la serie **converge**.

**Perché il criterio del rapporto "assorbe" facilmente il fattore polinomiale $n^2$, lasciando che sia il fattore esponenziale $2^n$ a decidere il risultato:** nel calcolo del rapporto, il termine polinomiale ha prodotto $\left(\frac{n+1}n\right)^2\to1$ (un fattore che "scompare" nel limite, perché il rapporto tra termini consecutivi di un polinomio tende sempre a $1$ — crescono, ma "lentamente" rispetto al rapporto stesso). È invece il fattore esponenziale $2^n$ a produrre il fattore costante $\frac12<1$ decisivo per la convergenza — coerente con quanto visto nel notebook 3.13: l'esponenziale "domina" qualunque potenza fissa, e qui è proprio il suo tasso di crescita (il denominatore $2$) a determinare l'esito.

````python
import sympy as sp

n = sp.symbols('n', positive=True, integer=True)
a_n = n**2 / 2**n

rapporto = sp.simplify(a_n.subs(n, n+1) / a_n)
print("a_(n+1)/a_n =", rapporto)

L = sp.limit(rapporto, n, sp.oo)
print("L =", L)

# Verifica con la somma parziale via SymPy
somma = sp.Sum(a_n, (n, 1, sp.oo)).doit()
print("\nSomma esatta della serie:", somma)
````

## Riepilogo finale

- **Criterio del confronto**: se $0\le a_n\le b_n$, la convergenza di $\sum b_n$ implica quella di $\sum a_n$; la divergenza di $\sum a_n$ implica quella di $\sum b_n$.
- **Criterio del rapporto**: $L=\lim\frac{a_{n+1}}{a_n}$; $L<1$ converge, $L>1$ diverge, $L=1$ non decide. Efficace con fattoriali e potenze.
- **Criterio della radice**: $L=\lim\sqrt[n]{a_n}$, stesse conclusioni del criterio del rapporto.
- Questi criteri permettono di stabilire la convergenza senza calcolare esplicitamente le somme parziali, uno strumento essenziale per le serie di potenze del prossimo notebook.
