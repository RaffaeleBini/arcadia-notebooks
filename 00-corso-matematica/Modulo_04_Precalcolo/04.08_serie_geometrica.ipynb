# 4.8 — Serie geometrica

**Prerequisiti:** 4.7 Serie numeriche, 3.17 Successioni (introduzione).


## Perché ci interessa

Nel notebook 3.17 abbiamo definito le successioni geometriche $a_n=a_0q^n$. Sommandole tutte, otteniamo la **serie geometrica** — probabilmente la serie più importante e più usata in tutta la matematica, dalla finanza (interesse composto, notebook 1.11) alla fisica, fino alle serie di potenze che vedremo nel notebook 4.11. A differenza della serie telescopica del notebook 4.7 (una tecnica ad hoc), qui otteniamo una **formula generale**, valida sempre sotto una condizione precisa.

## Teoria, passo dopo passo

### La formula della somma parziale

Consideriamo $\sum_{n=0}^\infty q^n = 1+q+q^2+q^3+\dots$ (il caso con $a_0=1$; il caso generale si ottiene moltiplicando tutto per $a_0$). Calcoliamo la somma parziale $S_N=\sum_{n=0}^N q^n$ con un trucco algebrico: moltiplichiamo $S_N$ per $q$ e sottraiamo:

$$S_N = 1+q+q^2+\dots+q^N$$
$$qS_N = \qquad q+q^2+\dots+q^N+q^{N+1}$$

Sottraendo membro a membro (quasi tutti i termini si cancellano, una struttura telescopica "nascosta", notebook 4.7):

$$S_N-qS_N = 1-q^{N+1} \quad\Rightarrow\quad S_N(1-q) = 1-q^{N+1}$$

Se $q\ne1$, dividiamo (invariantiva, notebook 0.5):

$$S_N = \frac{1-q^{N+1}}{1-q}$$

(Se $q=1$, la serie diventa banalmente $S_N=N+1\to\infty$: diverge sempre.)

### Il limite: quando la serie converge

Per trovare la somma della serie, calcoliamo $\lim_{N\to\infty}S_N$. Tutto dipende dal comportamento di $q^{N+1}$ (notebook 1.10-1.11, comportamento delle potenze):

- Se $|q|<1$: $q^{N+1}\to0$ (una potenza con base di modulo minore di 1 tende a zero, esattamente come $\left(\frac12\right)^N\to0$ visto nel notebook 4.7). Quindi $S_N\to\frac{1-0}{1-q}=\frac1{1-q}$: **la serie converge**.
- Se $|q|\ge1$ (e $q\ne1$, già trattato): $q^{N+1}$ non tende a zero (diverge in valore assoluto, o oscilla senza stabilizzarsi se $q\le-1$): **la serie diverge**.

### La formula finale

$$\sum_{n=0}^\infty q^n = \frac1{1-q} \qquad \text{se } |q|<1$$

Per la forma generale con $a_0$ qualsiasi:

$$\sum_{n=0}^\infty a_0q^n = \frac{a_0}{1-q} \qquad \text{se } |q|<1$$

Verifichiamo sull'esempio guida del notebook 4.7: $\sum\frac1{2^n}$ ha $a_0=1$, $q=\frac12$ (con $|q|<1$): somma $=\frac1{1-\frac12}=\frac1{1/2}=2$ — coincide con quanto avevamo intuito osservando le somme parziali.

````python
import sympy as sp

q, N = sp.symbols('q N', real=True)

# Verifica della formula della somma parziale
S_N = sp.Sum(q**sp.Symbol('n'), (sp.Symbol('n'), 0, N)).doit()
print("Somma parziale S_N (SymPy):", sp.simplify(S_N))

# Verifica numerica: q=1/2, confermiamo che la somma tende a 2
q_val = sp.Rational(1, 2)
for N_val in [5, 10, 20]:
    S = sum(q_val**k for k in range(N_val+1))
    print(f"N={N_val}: S_N = {float(S):.6f}")

print("\nFormula: 1/(1-q) =", 1/(1-q_val))

# Verifica per diversi valori di q: convergenza solo se |q|<1
casi = [sp.Rational(1,3), sp.Rational(2,3), sp.Rational(3,2), -1]
for q_test in casi:
    convergente = abs(q_test) < 1
    print(f"q={q_test}: |q|<1? {convergente}")
````

## Esercizio svolto e commentato

**Problema:** un farmaco viene assunto ogni giorno alla stessa dose $D$. Il corpo elimina il $30\%$ del farmaco presente ogni giorno (quindi ne resta il $70\%$). Qual è la quantità di farmaco nel corpo, a lungo termine, subito prima di ogni nuova dose (supponendo che il processo vada avanti indefinitamente)?

**Ragionamento:**

**Passo 1 — modelliamo il problema.** Chiamiamo $q=0.7$ (la frazione che resta ogni giorno, notebook 0.9). Il giorno $1$, subito prima della dose, nel corpo non c'è ancora nulla (la prima dose non è stata ancora presa in questo istante). Il giorno $2$, subito prima della seconda dose, resta il $70\%$ della prima dose: $Dq$. Il giorno $3$: la prima dose è passata per due "eliminazioni" ($Dq^2$), la seconda dose per una ($Dq$): totale $Dq+Dq^2$.

**Passo 2 — generalizziamo.** Al giorno $n+1$ (subito prima della dose $n+1$-esima), la quantità residua è la somma di tutte le dosi precedenti, ciascuna "invecchiata" di un numero diverso di giorni:

$$Q_n = Dq+Dq^2+\dots+Dq^n = D(q+q^2+\dots+q^n) = Dq\cdot\frac{1-q^n}{1-q}$$

(applicando la formula del notebook, con un piccolo adattamento per la somma che parte da $q^1$ invece che $q^0$: basta raccogliere un fattore $q$).

**Passo 3 — calcoliamo il limite per $n\to\infty$** (il comportamento "a lungo termine" richiesto): poiché $|q|=0.7<1$, $q^n\to0$ (teoria):

$$Q_\infty = Dq\cdot\frac{1-0}{1-q} = \frac{Dq}{1-q} = \frac{D(0.7)}{0.3} = \frac{7D}3 \approx 2.33D$$

**Conclusione**: a lungo termine, subito prima di ogni nuova dose, nel corpo resta circa $2.33$ volte la dose singola — un valore di **equilibrio** (o "steady state"), che il sistema raggiunge senza mai superarlo, indipendentemente da quante dosi siano già state prese.

**Perché la serie geometrica è il modello naturale per "accumulo con decadimento periodico" come questo:** ogni dose contribuisce con un fattore di decadimento diverso a seconda di quanto tempo è passato (esattamente come i termini $q,q^2,q^3,\dots$ di una serie geometrica), e la condizione $|q|<1$ (qui, il farmaco viene effettivamente eliminato, non accumulato all'infinito) è precisamente ciò che garantisce l'esistenza di un equilibrio finito — se il corpo eliminasse **meno** del $0\%$ per giorno (cioè $q\ge1$, un'eliminazione insufficiente), la quantità crescerebbe senza limite: un fenomeno di accumulo pericoloso, che la condizione di convergenza della serie geometrica identifica esattamente.

````python
import sympy as sp

D, q = sp.symbols('D q', positive=True)
q_val = sp.Rational(7, 10)

Q_infinito = D * q_val / (1 - q_val)
print("Quantità di equilibrio:", Q_infinito, "* D =", float(Q_infinito), "* D")

# Verifica numerica della convergenza di Q_n
D_val = 1
Q_n = lambda n: D_val * sum(q_val**k for k in range(1, n+1))
for n in [5, 10, 20, 50]:
    print(f"n={n}: Q_n = {float(Q_n(n)):.6f}")
````

## Riepilogo finale

- $\sum_{n=0}^\infty a_0q^n = \dfrac{a_0}{1-q}$ se $|q|<1$; la serie diverge se $|q|\ge1$.
- La formula si ricava con il "trucco" $S_N-qS_N$, che produce una cancellazione telescopica simile a quella del notebook 4.7.
- La serie geometrica modella naturalmente fenomeni di accumulo con decadimento a fattore costante (dosaggi di farmaci, interesse composto, rimbalzi con perdita di energia).
- La condizione $|q|<1$ non è un dettaglio tecnico: distingue tra un sistema che raggiunge un equilibrio finito e uno che accumula senza limite.
