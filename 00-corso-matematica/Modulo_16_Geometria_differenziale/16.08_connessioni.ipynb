## Perché ci interessa

Il notebook 16.3 aveva osservato un fatto scomodo: spazi tangenti $T_pM$ in punti diversi non hanno un'identificazione naturale. Questo crea un problema concreto: come si deriva un campo vettoriale su una varietà? In $\mathbb R^n$, per derivare un campo $V(x)$ lungo una direzione, sottraiamo $V$ in punti vicini e dividiamo per la distanza — ma questa sottrazione confronta $V(p+h)\in T_{p+h}$ con $V(p)\in T_p$, **due spazi vettoriali diversi** su una varietà generica, tra cui non c'è un modo canonico di sottrarre.

Una **connessione** risolve questo problema: fornisce un modo esplicito (e non canonico — è una scelta aggiuntiva) di "trasportare" un vettore tangente da un punto a un altro lungo una curva, permettendo così di derivare campi vettoriali. Su una varietà riemanniana, esiste una scelta di connessione particolarmente naturale, che risulta compatibile sia con la metrica sia con le geodetiche già incontrate: la **connessione di Levi-Civita**.

## Teoria, passo dopo passo

### La derivata covariante

Una **connessione affine** $\nabla$ associa a ogni campo vettoriale $X$ e ogni vettore tangente $v\in T_pM$ un nuovo vettore tangente $\nabla_vX\in T_pM$ (la "derivata di $X$ nella direzione $v$"), soddisfacendo proprietà di linearità in $v$, additività in $X$, e la regola di Leibniz $\nabla_v(fX)=v(f)X+f\,\nabla_vX$ per una funzione scalare $f$ (dove $v(f)$ è la derivata direzionale del notebook 16.3). In coordinate locali, $\nabla$ è determinata dai suoi valori sui campi di base $\partial_i=\partial/\partial x^i$:
$$\nabla_{\partial_i}\partial_j=\Gamma^k_{ij}\,\partial_k,$$
dove i coefficienti $\Gamma^k_{ij}$ — gli stessi **simboli di Christoffel** già comparsi nel notebook 16.7 senza essere definiti esplicitamente — determinano completamente $\nabla$ in coordinate, tramite la regola di Leibniz applicata a campi generici $X=X^i\partial_i$.

### La connessione di Levi-Civita

Tra tutte le possibili connessioni, ne esiste **una sola** che sia:
- **priva di torsione**: $\nabla_{\partial_i}\partial_j-\nabla_{\partial_j}\partial_i=0$ (cioè $\Gamma^k_{ij}=\Gamma^k_{ji}$, simmetrica nei due indici bassi);
- **compatibile con la metrica**: $v(g(X,Y))=g(\nabla_vX,Y)+g(X,\nabla_vY)$ per ogni $v,X,Y$ (la regola di Leibniz applicata al prodotto scalare $g$: derivare $g(X,Y)$ è "come" derivare un prodotto).

**Teorema (Levi-Civita).** Esiste un'unica connessione $\nabla$ su $(M,g)$ priva di torsione e compatibile con la metrica, data esplicitamente da
$$\Gamma^k_{ij}=\frac12g^{k\ell}\big(\partial_ig_{j\ell}+\partial_jg_{i\ell}-\partial_\ell g_{ij}\big),$$
dove $g^{k\ell}$ è la matrice inversa di $g_{ij}$ (esiste perché $g$ è definita positiva, quindi invertibile, notebook 6.16).

**Dimostrazione (schema).** Scrivendo la compatibilità con la metrica per le tre permutazioni cicliche degli indici $i,j,\ell$ (cioè $\partial_ig_{j\ell}=g(\nabla_{\partial_i}\partial_j,\partial_\ell)+g(\partial_j,\nabla_{\partial_i}\partial_\ell)$, e analoghe per $\partial_jg_{\ell i}$ e $\partial_\ell g_{ij}$), sommando le prime due equazioni e sottraendo la terza, i termini si riorganizzano — usando l'assenza di torsione per identificare $\nabla_{\partial_i}\partial_j=\nabla_{\partial_j}\partial_i$ — lasciando solo $2g(\nabla_{\partial_i}\partial_j,\partial_\ell)=\partial_ig_{j\ell}+\partial_jg_{i\ell}-\partial_\ell g_{ij}$. Moltiplicando per $g^{k\ell}$ (e sommando su $\ell$, usando $g^{k\ell}g_{\ell m}=\delta^k_m$, la definizione di matrice inversa) si isola $\Gamma^k_{ij}$ nella formula data. Questo mostra sia l'esistenza (la formula soddisfa le proprietà richieste, verificabile per sostituzione diretta) sia l'unicità (ogni connessione con quelle due proprietà deve soddisfare la stessa equazione, quindi avere gli stessi $\Gamma^k_{ij}$). $\blacksquare$

D'ora in avanti, quando si parla di "la" connessione su una varietà riemanniana, si intende sempre quella di Levi-Civita.

### Trasporto parallelo

Data una curva $\gamma(t)$ e un vettore tangente iniziale $v_0\in T_{\gamma(0)}M$, il **trasporto parallelo** di $v_0$ lungo $\gamma$ è l'unico campo vettoriale $V(t)\in T_{\gamma(t)}M$ lungo la curva che soddisfa $\nabla_{\dot\gamma}V=0$ (in coordinate, $\dot V^k+\Gamma^k_{ij}\dot x^iV^j=0$, un sistema lineare di EDO del primo ordine nelle componenti $V^k(t)$, con dato iniziale $V(0)=v_0$: esistenza e unicità garantite dalla teoria dei sistemi lineari di EDO — la versione lineare del teorema di esistenza e unicità del notebook 8.1, applicata al sistema di componenti $V^k(t)$) con $V(0)=v_0$. Intuitivamente, $V(t)$ è "il vettore che non cambia mai, secondo la connessione", il modo più coerente possibile di spostare un vettore tangente da uno spazio tangente all'altro lungo la curva — risolvendo esattamente il problema di identificazione tra $T_pM$ e $T_qM$ menzionato all'inizio.

**Le geodetiche, ridefinite.** Una curva $\gamma$ è una geodetica esattamente quando il suo **stesso** vettore velocità è trasportato parallelamente lungo sé stesso: $\nabla_{\dot\gamma}\dot\gamma=0$. In coordinate, questa condizione è $\ddot x^k+\Gamma^k_{ij}\dot x^i\dot x^j=0$ — **esattamente l'equazione della geodetica trovata nel notebook 16.7** con il metodo variazionale, ora ottenuta con il metodo, concettualmente diverso, del trasporto parallelo. Le due strade — minimizzare la lunghezza, e "non accelerare mai secondo la connessione" — portano alla stessa equazione: una coerenza non ovvia a priori, che conferma che la connessione di Levi-Civita è davvero l'oggetto giusto per formalizzare l'idea intuitiva di "andare dritti" su una superficie curva.

### Un esempio concreto: la sfera

Sulla sfera, il trasporto parallelo di un vettore lungo un cerchio massimo (una geodetica, notebook 16.7) lo lascia con lo stesso angolo rispetto alla curva; lungo un parallelo (che **non** è una geodetica, tranne l'equatore), il trasporto parallelo produce un vettore che, dopo aver percorso l'intero parallelo, **non torna al vettore di partenza**, ma è ruotato di un angolo — l'**olonomia** del trasporto parallelo, un fenomeno che nella teoria della curvatura (notebook 16.9) risulterà essere direttamente collegato alla curvatura racchiusa dal cammino.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Trasporto parallelo lungo un parallelo della sfera a theta = theta0 (non geodetica, tranne
# l'equatore theta0 = pi/2), usando i simboli di Christoffel della metrica sferica.
# V = (V_theta, V_phi) le componenti del vettore trasportato, funzione di t=phi.

theta0 = np.pi/4  # un parallelo a 45 gradi, non l'equatore

def Gamma_theta_phiphi(theta):
    return -np.sin(theta)*np.cos(theta)

def Gamma_phi_thetaphi(theta):
    return np.cos(theta)/np.sin(theta)

def trasporto_parallelo(t, V):
    # lungo il parallelo: dtheta/dt = 0, dphi/dt = 1 (t = phi stesso)
    V_theta, V_phi = V
    dV_theta = -Gamma_theta_phiphi(theta0) * V_phi          # dtheta/dt=0, dphi/dt=1
    dV_phi = -Gamma_phi_thetaphi(theta0) * V_theta          # solo il termine con dphi/dt=1 sopravvive
    return [dV_theta, dV_phi]

V0 = [1.0, 0.0]  # vettore iniziale puramente radiale (direzione theta)
sol = solve_ivp(trasporto_parallelo, [0, 2*np.pi], V0, dense_output=True, max_step=0.001)

V_finale = sol.y[:, -1]
print(f"Vettore trasportato dopo un giro completo del parallelo a theta={theta0:.3f}:")
print(f"  V_iniziale = {V0}")
print(f"  V_finale   = {V_finale}")

angolo_rotazione = np.arctan2(V_finale[1]*np.sin(theta0), V_finale[0])
print(f"Angolo di olonomia (approssimato): {np.degrees(angolo_rotazione):.2f} gradi")
print(f"Angolo atteso per un parallelo, 2*pi*cos(theta0) in radianti: {np.degrees(2*np.pi*np.cos(theta0)):.2f} gradi")
```

## Esercizio svolto e commentato

**Problema:** verifica esplicitamente, per la metrica euclidea $g_{ij}=\delta_{ij}$ su $\mathbb R^n$, che la connessione di Levi-Civita ha tutti i simboli di Christoffel nulli, usando la formula esplicita.

**Ragionamento:**

**Passo 1 — scriviamo la formula generale.** $\Gamma^k_{ij}=\dfrac12g^{k\ell}(\partial_ig_{j\ell}+\partial_jg_{i\ell}-\partial_\ell g_{ij})$.

**Passo 2 — osserviamo che $g_{ij}=\delta_{ij}$ è costante.** Ogni derivata parziale di una funzione costante è zero: $\partial_ig_{j\ell}=\partial_jg_{i\ell}=\partial_\ell g_{ij}=0$ per ogni scelta di indici.

**Passo 3 — concludiamo.** $\Gamma^k_{ij}=\dfrac12g^{k\ell}(0+0-0)=0$ per ogni $i,j,k$.

**Conclusione:** questo conferma, ora con la formula esplicita della connessione di Levi-Civita (non solo per il ragionamento intuitivo del notebook 16.7), che in coordinate cartesiane euclidee i simboli di Christoffel si annullano identicamente — motivo per cui l'equazione della geodetica si riduce a $\ddot x^k=0$ e le geodetiche sono rette, e per cui il trasporto parallelo in $\mathbb R^n$ euclideo coincide semplicemente con "tenere le componenti costanti", senza nessuna correzione dovuta alla curvatura dello spazio (che infatti, come vedremo nel notebook 16.9, è nulla).

```python
import sympy as sp

n = 3
x = sp.symbols(f'x1:{n+1}', real=True)
g = sp.eye(n)  # metrica euclidea, costante
g_inv = g.inv()

def christoffel(k, i, j):
    s = 0
    for l in range(n):
        s += g_inv[k, l] * (sp.diff(g[j, l], x[i]) + sp.diff(g[i, l], x[j]) - sp.diff(g[i, j], x[l]))
    return sp.simplify(s / 2)

tutti_nulli = True
for k in range(n):
    for i in range(n):
        for j in range(n):
            val = christoffel(k, i, j)
            if val != 0:
                tutti_nulli = False
                print(f"Gamma^{k}_{{{i}{j}}} = {val}  <-- non nullo!")

print("Tutti i simboli di Christoffel sono nulli per la metrica euclidea:", tutti_nulli)
```

Su una varietà curva, i simboli di Christoffel **non nulli** introducono una correzione a ogni istante che dipende dalla curva percorsa: se la curva racchiude una regione dove la curvatura non è nulla (come qualunque parallelo non-equatoriale della sfera, che racchiude una calotta), le correzioni si accumulano in modo da non cancellarsi esattamente al ritorno al punto di partenza, producendo l'angolo di olonomia osservato numericamente nella cella di verifica.

**3.** L'equazione $\dot V^k=-\Gamma^k_{ij}(\gamma(t))\dot x^i(t)V^j(t)$ è lineare in $V^j$ (i coefficienti $\Gamma^k_{ij}(\gamma(t))\dot x^i(t)$ dipendono solo da $t$, essendo $\gamma(t)$ già nota e fissata, non da $V$): un sistema di EDO lineari del primo ordine. La teoria dei sistemi lineari di EDO (generalizzazione del teorema di esistenza e unicità del notebook 8.1) garantisce che le soluzioni di un sistema lineare esistono per **tutto** l'intervallo su cui i coefficienti sono definiti e continui (qui, tutto il dominio della curva $\gamma$) — a differenza di sistemi non lineari come l'equazione della geodetica (dove $\Gamma^k_{ij}\dot x^i\dot x^j$ è quadratico in $\dot x$, quindi non lineare nelle incognite $x^k$ stesse), per cui il teorema di esistenza e unicità garantisce solo una soluzione **locale**, potenzialmente non estendibile a tutto l'intervallo (le soluzioni potrebbero "scappare all'infinito" in tempo finito, un fenomeno tipico delle equazioni non lineari).

**4.** $\dfrac{d}{dt}g(V,W)=g(\nabla_{\dot\gamma}V,W)+g(V,\nabla_{\dot\gamma}W)$ per compatibilità della connessione con la metrica (proprietà che definisce Levi-Civita). Se $V,W$ sono entrambi trasportati parallelamente, $\nabla_{\dot\gamma}V=0$ e $\nabla_{\dot\gamma}W=0$ per definizione, quindi $\dfrac{d}{dt}g(V,W)=g(0,W)+g(V,0)=0$: $g(V,W)$ è costante. In particolare, prendendo $W=V$: $g(V,V)=\|V\|^2$ è costante, cioè **il trasporto parallelo preserva la lunghezza dei vettori** (e, più in generale, gli angoli tra vettori trasportati) — coerente con l'idea intuitiva che il trasporto parallelo sposti un vettore "senza deformarlo", solo cambiandone l'orientamento relativo allo spazio tangente locale.

## Riepilogo finale

- Una **connessione** $\nabla$ permette di derivare campi vettoriali su una varietà, risolvendo il problema (notebook 16.3) che spazi tangenti in punti diversi non sono canonicamente identificabili.
- Su una varietà riemanniana esiste un'unica connessione — quella di **Levi-Civita** — priva di torsione e compatibile con la metrica, con simboli di Christoffel $\Gamma^k_{ij}$ espressi esplicitamente in termini di $g_{ij}$ e delle sue derivate.
- Il **trasporto parallelo** sposta un vettore tangente lungo una curva mantenendolo "il più costante possibile" secondo la connessione; preserva lunghezze e angoli (compatibilità con la metrica).
- Le geodetiche, definite nel notebook 16.7 tramite minimizzazione della lunghezza, coincidono esattamente con le curve il cui vettore velocità è trasportato parallelamente lungo sé stesso: due definizioni concettualmente diverse che portano alla stessa equazione.
- Il trasporto parallelo lungo curve chiuse può produrre olonomia (un vettore che non torna a sé stesso): un fenomeno strettamente legato alla curvatura, il tema del prossimo notebook.
