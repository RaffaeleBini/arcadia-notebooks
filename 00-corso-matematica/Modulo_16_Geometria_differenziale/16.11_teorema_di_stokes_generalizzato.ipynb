## Teoria, passo dopo passo

### Varietà con bordo e orientazione

Per enunciare il teorema serve una nozione tecnica: una **varietà con bordo** $M$ è come una varietà (notebook 16.5), ma alcune carte hanno per immagine un semispazio $\{x_n\ge0\}\subset\mathbb R^n$ invece di tutto $\mathbb R^n$; i punti che corrispondono a $x_n=0$ formano il **bordo** $\partial M$, che è a sua volta una varietà (senza bordo) di dimensione $n-1$. Un intervallo $[a,b]$ è una varietà con bordo di dimensione 1, con $\partial[a,b]=\{a,b\}$; un disco chiuso è una varietà con bordo di dimensione 2, con bordo la circonferenza; una palla chiusa in $\mathbb R^3$ ha come bordo la sfera.

Serve anche che $M$ sia **orientabile** (esista una scelta coerente di "verso" su ogni spazio tangente, che varia con continuità), e che il bordo $\partial M$ riceva un'**orientazione indotta** (convenzionalmente: il verso normale uscente "seguito da" l'orientazione del bordo riproduce l'orientazione di $M$ — per una regione piana, questo dà il verso antiorario sul bordo se la regione ha l'orientazione standard, coerente con la convenzione già usata implicitamente nel notebook 7.23).

### L'integrale di una forma su una varietà orientata

Data una $k$-forma $\omega$ (notebook 16.4) su una varietà orientata $M$ di dimensione $k$, il suo integrale $\int_M\omega$ si definisce carta per carta: in ogni carta, $\omega$ diventa (a meno del segno, dipendente dall'orientazione) $f(x)\,dx_1\wedge\cdots\wedge dx_k$ per una funzione $f$, e si pone $\int_M\omega=\int f(x)\,dx_1\cdots dx_k$, l'ordinario integrale multiplo (notebook 7.14-7.16) — un integrale ben definito, indipendente dalla carta, esattamente grazie alla presenza del prodotto esterno antisimmetrico nella definizione di forma, che assorbe automaticamente il fattore Jacobiano del cambiamento di variabili (notebook 7.15) nel cambio di carta.

### Il teorema di Stokes generalizzato

**Teorema.** Sia $M$ una varietà orientata con bordo, di dimensione $n$, con bordo $\partial M$ orientato in modo indotto, e sia $\omega$ una $(n-1)$-forma con supporto compatto su $M$. Allora
$$\int_M d\omega=\int_{\partial M}\omega.$$

Questo enunciato — un'unica identità, valida in qualunque dimensione $n$ — è il vero "teorema di Stokes": tutto ciò che serve fare per ottenere i teoremi classici del Modulo 7 è scegliere $n$, la forma $\omega$, e tradurre il linguaggio delle forme in quello dei campi vettoriali usando la corrispondenza del notebook 16.4 (gradiente/rotore/divergenza come casi di $d$).

### I casi particolari

- **$n=1$, $M=[a,b]$: il teorema fondamentale del calcolo.** Con $\omega=f$ una 0-forma (funzione), $d\omega=f'(x)\,dx$ una 1-forma. $\partial M=\{a,b\}$ con orientazione indotta $+b,-a$ (il bordo "destro" con segno positivo, quello "sinistro" con segno negativo — la convenzione standard per intervalli). Il teorema dà $\int_a^bf'(x)\,dx=f(b)-f(a)$: **esattamente il teorema fondamentale del calcolo** (notebook 5.16), qui riconosciuto come il caso $n=1$ del teorema generale.
- **$n=2$, $M$ una regione piana: il teorema di Green.** Con $\omega=P\,dx+Q\,dy$ una 1-forma, $d\omega=(Q_x-P_y)\,dx\wedge dy$ (notebook 16.4). Il teorema dà $\displaystyle\iint_M(Q_x-P_y)\,dx\,dy=\oint_{\partial M}P\,dx+Q\,dy$ — **esattamente il teorema di Green** (notebook 7.23).
- **$n=2$ (come sottovarietà di $\mathbb R^3$): il teorema di Stokes classico.** Con $\omega$ la 1-forma associata al campo $F=(P,Q,R)$ (notebook 7.20) e $M$ una superficie orientata in $\mathbb R^3$, $d\omega$ è la 2-forma associata a $\mathrm{rot}\,F$ (notebook 16.4). Il teorema dà $\displaystyle\iint_M(\mathrm{rot}\,F)\cdot n\,dA=\oint_{\partial M}F\cdot d\gamma$ — **esattamente il teorema di Stokes** (notebook 7.25), di cui il teorema di Green è, non a caso, il caso particolare in cui la superficie è piatta (con normale $n=(0,0,1)$ costante, $\mathrm{rot}\,F\cdot n$ si riduce esattamente a $Q_x-P_y$).
- **$n=3$: il teorema della divergenza.** Con $\omega$ la 2-forma associata a $F=(P,Q,R)$ (nel senso "di flusso" del notebook 16.4) e $M$ un solido in $\mathbb R^3$, $d\omega$ è la 3-forma associata a $\mathrm{div}\,F$. Il teorema dà $\displaystyle\iiint_M(\mathrm{div}\,F)\,dV=\oiint_{\partial M}F\cdot n\,dA$ — **esattamente il teorema della divergenza** (notebook 7.24).

### Perché $d^2=0$ e "il bordo del bordo è vuoto" sono la stessa cosa

Un fatto elegante conclude il modulo: applicando il teorema generale due volte, a $\omega$ su $M$ e poi (formalmente) considerando che $\partial(\partial M)=\emptyset$ sempre (il bordo di una varietà con bordo non ha a sua volta un bordo — per esempio il bordo del bordo di una palla, la sfera, non ha bordo), si ottiene $\int_{\partial(\partial M)}\omega=0=\int_{\partial M}d\omega=\int_Md(d\omega)$: coerente con l'identità algebrica $d^2=0$ già dimostrata nel notebook 16.4. Questa non è una coincidenza: $d^2=0$ e "il bordo del bordo è vuoto" sono, in un senso preciso reso rigoroso dalla teoria della coomologia (che esula da questo corso introduttivo), **due facce della stessa dualità** tra forme differenziali e catene di integrazione.

```python
import numpy as np
from scipy import integrate

# Verifica numerica del teorema di Stokes generalizzato nel caso n=2 (Green),
# confrontando l'integrale doppio di (Qx - Py) con l'integrale di linea sul bordo,
# per un campo F=(P,Q) su un disco di raggio R centrato nell'origine.

def P(x, y):
    return -y**2
def Q(x, y):
    return x*y

Rdisco = 1.5

# Lato sinistro: integrale doppio di (Q_x - P_y) sul disco
def Qx_meno_Py(x, y):
    Qx = y          # d/dx (x*y) = y
    Py = -2*y       # d/dy (-y^2) = -2y
    return Qx - Py  # = y - (-2y) = 3y

def integrando_polare(rho, theta):
    x, y = rho*np.cos(theta), rho*np.sin(theta)
    return Qx_meno_Py(x, y) * rho  # Jacobiano delle coordinate polari

lato_sinistro, _ = integrate.dblquad(integrando_polare, 0, 2*np.pi, 0, Rdisco)
print(f"Integrale doppio di (Qx - Py) sul disco: {lato_sinistro:.6f}")

# Lato destro: integrale di linea sul bordo (la circonferenza di raggio Rdisco)
def integrando_linea(theta):
    x, y = Rdisco*np.cos(theta), Rdisco*np.sin(theta)
    dx_dtheta, dy_dtheta = -Rdisco*np.sin(theta), Rdisco*np.cos(theta)
    return P(x, y)*dx_dtheta + Q(x, y)*dy_dtheta

lato_destro, _ = integrate.quad(integrando_linea, 0, 2*np.pi)
print(f"Integrale di linea di P dx + Q dy sul bordo: {lato_destro:.6f}")

print(f"Coincidono (teorema di Green, caso n=2 del teorema generale): {np.isclose(lato_sinistro, lato_destro)}")
```

## Esercizio svolto e commentato

**Problema:** verifica il teorema di Stokes generalizzato nel caso $n=3$ (teorema della divergenza), per il campo $F=(x,y,z)$ (il campo radiale) sulla palla di raggio $R$, confrontando $\iiint_M\mathrm{div}\,F\,dV$ con $\oiint_{\partial M}F\cdot n\,dA$.

**Ragionamento:**

**Passo 1 — calcoliamo la divergenza.** $\mathrm{div}\,F=\partial_x(x)+\partial_y(y)+\partial_z(z)=1+1+1=3$ (costante).

**Passo 2 — calcoliamo il lato sinistro (integrale di volume).** $\displaystyle\iiint_M3\,dV=3\cdot\mathrm{Vol}(M)=3\cdot\frac43\pi R^3=4\pi R^3$ (volume della palla, notebook 7.16, in coordinate sferiche).

**Passo 3 — calcoliamo il lato destro (integrale di superficie).** Sulla sfera $\partial M$ di raggio $R$, il normale uscente è $n=(x,y,z)/R$ (il raggio normalizzato, come nel notebook 16.9). Quindi $F\cdot n=(x,y,z)\cdot(x,y,z)/R=(x^2+y^2+z^2)/R=R^2/R=R$ (usando $x^2+y^2+z^2=R^2$ sulla sfera di raggio $R$) — **costante** su tutta la sfera.

**Passo 4 — integriamo la costante $R$ sulla sfera.** $\displaystyle\oiint_{\partial M}F\cdot n\,dA=R\cdot\mathrm{Area}(\partial M)=R\cdot4\pi R^2=4\pi R^3$ (area della sfera, notebook 16.2, esercizio svolto).

**Conclusione:** entrambi i lati danno $4\pi R^3$ — il teorema della divergenza è verificato su questo esempio esplicito, ed essendo entrambi i calcoli riconducibili, tramite le corrispondenze del notebook 16.4, all'unico teorema di Stokes generalizzato con $\omega$ la 2-forma associata a $F$ e $d\omega$ la 3-forma associata a $\mathrm{div}\,F$, questo esempio concreto è anche una verifica (in un caso particolare) del teorema unificante enunciato nella teoria.

```python
import numpy as np
from scipy import integrate

R = 2.0

# Lato sinistro: integrale di volume della divergenza (costante = 3)
volume_palla = (4/3)*np.pi*R**3
lato_sinistro = 3 * volume_palla
print(f"Integrale di volume di div(F)=3: {lato_sinistro:.6f}")

# Lato destro: integrale di superficie di F.n sulla sfera di raggio R (F.n = R costante)
area_sfera = 4*np.pi*R**2
lato_destro = R * area_sfera
print(f"Integrale di superficie di F.n: {lato_destro:.6f}")

print(f"Coincidono (teorema della divergenza, caso n=3): {np.isclose(lato_sinistro, lato_destro)}")
print(f"Valore comune atteso 4*pi*R^3 = {4*np.pi*R**3:.6f}")
```

## Riepilogo finale

- Il **teorema di Stokes generalizzato**, $\int_Md\omega=\int_{\partial M}\omega$, è un unico enunciato valido in qualunque dimensione, per varietà orientate con bordo.
- Scegliendo la dimensione e il grado della forma opportunamente, questo teorema **contiene** come casi particolari il teorema fondamentale del calcolo ($n=1$), il teorema di Green ($n=2$ nel piano), il teorema di Stokes classico ($n=2$ in $\mathbb R^3$) e il teorema della divergenza ($n=3$) — quattro risultati del Modulo 5 e del Modulo 7 che sembravano indipendenti.
- L'identità algebrica $d^2=0$ (notebook 16.4) e il fatto geometrico "il bordo del bordo è vuoto" sono manifestazioni collegate della stessa dualità tra forme e integrazione.
