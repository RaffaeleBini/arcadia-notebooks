**Prerequisiti:** 16.8 Connessioni, 16.2 Superfici (prima forma fondamentale), 6.17-6.19 Autovalori, autovettori e diagonalizzazione.


## Perché ci interessa

Il notebook 16.1 aveva già una nozione di curvatura per le curve: $\kappa=\|\gamma''(s)\|$, quanto la curva si scosta dall'essere retta. Il notebook 16.8 ha mostrato un fenomeno nuovo: trasportando parallelamente un vettore lungo un cammino chiuso su una superficie curva, il vettore può tornare **ruotato** — l'olonomia. Questo notebook formalizza la curvatura per superfici e varietà, mostrando che l'olonomia non è un fenomeno a sé, ma è **esattamente misurata** da un oggetto chiamato tensore di curvatura di Riemann.

Il risultato più sorprendente sarà il **Theorema Egregium** di Gauss: la curvatura di una superficie, definita apparentemente usando come è immersa in $\mathbb R^3$ (le due direzioni principali in cui si piega), è in realtà calcolabile **solo dalla prima forma fondamentale** — un dato puramente intrinseco. Due superfici con la stessa geometria interna (stessa metrica) hanno la stessa curvatura, anche se sono piegate diversamente nello spazio ambiente.

## Teoria, passo dopo passo

### Il tensore di curvatura di Riemann

Sulla sfera, trasportare parallelamente un vettore lungo due cammini diversi da $p$ a $q$ può dare risultati diversi in $q$: il trasporto parallelo **non commuta**. Il tensore di curvatura di Riemann misura precisamente questa non commutatività, tramite le derivate covarianti (notebook 16.8):
$$R(X,Y)Z=\nabla_X\nabla_YZ-\nabla_Y\nabla_XZ-\nabla_{[X,Y]}Z,$$
dove $[X,Y]$ è il commutatore dei campi vettoriali (il termine che compensa il fatto che $X,Y$ stessi potrebbero non "commutare" come direzioni di derivazione, un dettaglio tecnico che si annulla se $X=\partial_i,Y=\partial_j$ sono campi coordinati, il caso che useremo esclusivamente in questo notebook). In coordinate, $R^\ell_{kij}$ (le componenti di $R(\partial_i,\partial_j)\partial_k=R^\ell_{kij}\partial_\ell$) si esprimono in termini dei simboli di Christoffel e delle loro derivate prime:
$$R^\ell_{kij}=\partial_i\Gamma^\ell_{jk}-\partial_j\Gamma^\ell_{ik}+\Gamma^\ell_{im}\Gamma^m_{jk}-\Gamma^\ell_{jm}\Gamma^m_{ik}.$$

**Fatto chiave.** $R\equiv0$ se e solo se $\Gamma^k_{ij}\equiv0$ in un intorno (più precisamente, se e solo se esistono coordinate locali in cui la metrica ha $g_{ij}=\delta_{ij}$ costante): $R$ misura esattamente quanto una varietà riemanniana si scosta dall'essere localmente indistinguibile dallo spazio euclideo. In $\mathbb R^n$ con la metrica standard, $\Gamma\equiv0$ (notebook 16.8), quindi $R\equiv0$: nessuna curvatura, coerentemente con l'intuizione.

### Curvatura sezionale e curvatura di Gauss

Per una superficie ($n=2$), il tensore di curvatura ha, per ragioni di simmetria algebriche di $R$ (antisimmetria negli ultimi due indici, tra le altre), essenzialmente **una sola componente indipendente**, che si condensa nella **curvatura di Gauss**
$$K=\frac{R_{1212}}{g_{11}g_{22}-g_{12}^2},\qquad R_{1212}=g_{1\ell}R^\ell_{212}.$$

### Il Theorema Egregium

**Teorema (Gauss, 1827).** Per una superficie $\Sigma\subset\mathbb R^3$, la curvatura di Gauss $K$, calcolata dalla formula sopra usando solo $E,F,G$ (la prima forma fondamentale, notebook 16.2) e le loro derivate, **coincide** con il prodotto delle **curvature principali** $\kappa_1,\kappa_2$ — gli autovalori (notebook 6.17-6.18) dell'**operatore forma** $S_p:T_p\Sigma\to T_p\Sigma$, definito da $S_p(v)=-\nabla_vN$ (la derivata del campo normale unitario $N$ nella direzione $v$, che risulta tangente alla superficie, un fatto verificabile derivando $N\cdot N=1$), un oggetto definito usando esplicitamente **come la superficie è immersa** in $\mathbb R^3$ (la sua "seconda forma fondamentale" $\mathrm{II}(v,w)=g(S_p(v),w)$, un'altra forma quadratica, notebook 6.23, come $\mathrm I$, ma stavolta che misura quanto la superficie si piega, non le sue lunghezze interne).

**Perché è "egregium" (straordinario).** A priori, $K=\kappa_1\kappa_2$ dipende da $S_p$, che dipende da come $\Sigma$ sta in $\mathbb R^3$ (**estrinseco**): due superfici con la stessa forma fondamentale ma piegate diversamente potrebbero avere operatori forma diversi. Il teorema di Gauss dice che, nonostante questo, $K$ è determinabile **solo** da $E,F,G$ (**intrinseco**) — quindi $K$ è la stessa per due superfici isometriche (che preservano $\mathrm I$), anche se le loro forme nello spazio ambiente sono completamente diverse. Questo è esattamente il tipo di risultato anticipato nel notebook 16.1 con il teorema fondamentale delle curve: separare le proprietà "vere" (intrinseche) da quelle che dipendono dall'immersione — solo che qui il risultato è molto più sorprendente, perché $\kappa_1,\kappa_2$ presi singolarmente **sono** estrinseci (dipendono da come la superficie è piegata), eppure il loro **prodotto** non lo è.

**Conseguenza pratica.** Il cilindro ha $\kappa_1=1/r$ (si piega in una direzione) e $\kappa_2=0$ (è dritto nella direzione $v$, notebook 16.2), quindi $K=\kappa_1\kappa_2=0$: **stessa curvatura di Gauss del piano** ($K=0$), coerente con il fatto già osservato nel notebook 16.2 e 16.6 che cilindro e piano hanno la stessa prima forma fondamentale (si ottiene uno dall'altro arrotolando un foglio senza stirarlo — un'isometria locale). La sfera, invece, ha $K=1/r^2>0$ ovunque, e **nessuna** deformazione senza stiramento può appiattirla: questo è il motivo geometrico rigoroso per cui non esistono mappe geografiche piatte che preservino tutte le distanze — un fatto che, prima del Theorema Egregium, era solo un'osservazione empirica dei cartografi.

### Curvatura di Ricci e curvatura scalare (cenno per dimensioni superiori)

In dimensione $n>2$, il tensore $R^\ell_{kij}$ ha molte più componenti indipendenti. Due contrazioni utili (somme su indici ripetuti, che riducono il numero di indici, un'operazione analoga alla traccia di una matrice, notebook 6.17) sono il **tensore di Ricci** $\mathrm{Ric}_{ij}=R^k_{ikj}$ e la **curvatura scalare** $S=g^{ij}\mathrm{Ric}_{ij}$, un singolo numero per punto che generalizza $K$ (per $n=2$, $S=2K$). Questi oggetti, centrali nella relatività generale, esulano dallo scopo di questo modulo introduttivo, ma la loro costruzione segue esattamente lo stesso principio: contrarre il tensore di curvatura completo per ottenere informazioni più sintetiche.

```python
import sympy as sp

theta, phi, r = sp.symbols('theta phi r', positive=True)
g = sp.diag(r**2, r**2*sp.sin(theta)**2)  # metrica sferica in (theta, phi)
coords = [theta, phi]
n = 2
g_inv = g.inv()

def christoffel_simbolo(k, i, j):
    s = 0
    for l in range(n):
        s += g_inv[k, l] * (sp.diff(g[j, l], coords[i]) + sp.diff(g[i, l], coords[j]) - sp.diff(g[i, j], coords[l]))
    return sp.simplify(s / 2)

Gamma = [[[christoffel_simbolo(k, i, j) for j in range(n)] for i in range(n)] for k in range(n)]

def riemann_su(l, k, i, j):
    termine = sp.diff(Gamma[l][j][k], coords[i]) - sp.diff(Gamma[l][i][k], coords[j])
    for m in range(n):
        termine += Gamma[l][i][m]*Gamma[m][j][k] - Gamma[l][j][m]*Gamma[m][i][k]
    return sp.simplify(termine)

R_1212 = sum(g[0, l] * riemann_su(l, 1, 0, 1) for l in range(n))
R_1212 = sp.simplify(R_1212)
det_g = sp.simplify(g[0,0]*g[1,1] - g[0,1]*g[1,0])
K = sp.simplify(R_1212 / det_g)

print("R_1212 =", R_1212)
print("det(g) =", det_g)
print("Curvatura di Gauss K =", K, " (attesa: 1/r^2)")
```

## Esercizio svolto e commentato

**Problema:** verifica, tramite l'operatore forma, che $K=\kappa_1\kappa_2=1/r^2$ per la sfera di raggio $r$ (il calcolo "estrinseco"), da confrontare con il calcolo "intrinseco" appena fatto con SymPy.

**Ragionamento:**

**Passo 1 — troviamo il campo normale.** Per la sfera $\sigma(\theta,\phi)=r(\sin\theta\cos\phi,\sin\theta\sin\phi,\cos\theta)$, scegliamo il vettore normale unitario **entrante** $N=-\sigma/r=-(\sin\theta\cos\phi,\sin\theta\sin\phi,\cos\theta)$ (il raggio, invertito e diviso per la sua lunghezza $r$: per una sfera centrata nell'origine, il normale è sempre radiale; questa scelta di segno è solo una convenzione, che rende le curvature principali positive).

**Passo 2 — calcoliamo l'operatore forma sui vettori di base.** $\partial_\theta N=-(\cos\theta\cos\phi,\cos\theta\sin\phi,-\sin\theta)=-\frac1r\sigma_\theta$ (confrontando con $\sigma_\theta=r(\cos\theta\cos\phi,\cos\theta\sin\phi,-\sin\theta)$, notebook 16.2). Quindi $S(\sigma_\theta)=-\partial_\theta N=\frac1r\sigma_\theta$: $\sigma_\theta$ è un **autovettore** di $S$ con autovalore $\kappa_1=1/r$.

**Passo 3 — ripetiamo per $\sigma_\phi$.** Un conto analogo (che omettiamo per brevità, identico nella struttura al passo 2) dà $S(\sigma_\phi)=\frac1r\sigma_\phi$: $\sigma_\phi$ è anch'esso autovettore, con lo stesso autovalore $1/r$.

**Passo 4 — troviamo le curvature principali.** Essendo $\sigma_\theta,\sigma_\phi$ una base di autovettori (notebook 6.18) con lo stesso autovalore $1/r$: $\kappa_1=\kappa_2=1/r$ (la sfera è un punto **ombelicale** ovunque: si piega ugualmente in ogni direzione, un fatto specifico della sfera che non vale per superfici generiche).

**Passo 5 — calcoliamo $K$.** $K=\kappa_1\kappa_2=\dfrac1r\cdot\dfrac1r=\dfrac1{r^2}$ — **esattamente il valore trovato con il calcolo intrinseco** nella cella di codice precedente, che usava solo $E,F,G$ senza mai menzionare l'operatore forma o il vettore normale. Questa coincidenza (che è, precisamente, il contenuto del Theorema Egregium) non è un caso: è il teorema di Gauss verificato su un esempio concreto.

```python
import sympy as sp

theta, phi, r = sp.symbols('theta phi r', positive=True)
sigma = sp.Matrix([r*sp.sin(theta)*sp.cos(phi), r*sp.sin(theta)*sp.sin(phi), r*sp.cos(theta)])
N = -sigma / r  # normale interno (convenzione che da' curvature principali positive)

sigma_theta = sp.diff(sigma, theta)
sigma_phi = sp.diff(sigma, phi)
N_theta = sp.diff(N, theta)
N_phi = sp.diff(N, phi)

# S(sigma_theta) = -dN/dtheta ; verifichiamo che e' proporzionale a sigma_theta
S_sigma_theta = -N_theta
rapporto_theta = sp.simplify(S_sigma_theta[0] / sigma_theta[0])
print("S(sigma_theta) / sigma_theta =", rapporto_theta, " (curvatura principale kappa_1)")

S_sigma_phi = -N_phi
rapporto_phi = sp.simplify(S_sigma_phi[1] / sigma_phi[1])
print("S(sigma_phi) / sigma_phi =", rapporto_phi, " (curvatura principale kappa_2)")

K_estrinseco = sp.simplify(rapporto_theta * rapporto_phi)
print("K = kappa_1 * kappa_2 =", K_estrinseco, " (confrontare con 1/r^2 del calcolo intrinseco)")
```

## Riepilogo finale

- Il **tensore di curvatura di Riemann** $R$ misura la non commutatività delle derivate covarianti, e si annulla identicamente esattamente quando la metrica è, localmente, quella euclidea.
- Per le superfici, $R$ si condensa nella **curvatura di Gauss** $K$, esprimibile come prodotto delle curvature principali $\kappa_1,\kappa_2$ (autovalori dell'operatore forma, un oggetto estrinseco).
- Il **Theorema Egregium** di Gauss mostra che $K$, nonostante sia definito tramite un oggetto estrinseco, è in realtà calcolabile solo dalla prima forma fondamentale: è una quantità **intrinseca**.
- Conseguenza: superfici con $K$ diverso (come sfera, $K>0$, e piano, $K=0$) non possono mai essere isometriche, nemmeno localmente — da cui l'impossibilità di mappe geografiche perfette.
- In dimensioni superiori, contrazioni del tensore di Riemann (Ricci, curvatura scalare) generalizzano $K$, con applicazioni centrali in relatività generale.
