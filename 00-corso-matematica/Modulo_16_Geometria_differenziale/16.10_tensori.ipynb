## Perché ci interessa

Nei notebook precedenti abbiamo incontrato molti oggetti che si comportano in modo simile ma sono stati introdotti separatamente: vettori tangenti (notebook 16.3), covettori e forme differenziali (notebook 16.4), la metrica $g_{ij}$ (notebook 16.6), il tensore di curvatura $R^\ell_{kij}$ (notebook 16.9). Tutti hanno una cosa in comune: sono definiti da componenti numeriche che **cambiano in modo prevedibile** quando si cambia base o sistema di coordinate.

Questo notebook fornisce il linguaggio unificante — i **tensori** — che inquadra tutti questi oggetti come casi particolari di un'unica struttura algebrica. Il vantaggio non è solo di sistemazione: capire la legge di trasformazione generale dei tensori permette di riconoscere immediatamente se una quantità scritta in componenti ha un significato geometrico intrinseco (è un tensore) o dipende accidentalmente dalla scelta di coordinate (non lo è) — una distinzione cruciale, per esempio, per i simboli di Christoffel del notebook 16.8, che **non sono le componenti di un tensore**, come vedremo.

## Teoria, passo dopo passo

### Tensori come applicazioni multilineari

Fissato uno spazio vettoriale $V$ di dimensione $n$ (pensiamo a $V=T_pM$, notebook 16.3) e il suo duale $V^*$ (notebook 16.4), un **tensore di tipo $(r,s)$** è un'applicazione multilineare
$$T:\underbrace{V^*\times\cdots\times V^*}_{r}\times\underbrace{V\times\cdots\times V}_{s}\to\mathbb R$$
(lineare in ciascuno degli $r+s$ argomenti separatamente, generalizzando la linearità di un'applicazione lineare, notebook 6.13, a più argomenti). L'insieme dei tensori di tipo $(r,s)$ su $V$, con le operazioni naturali di somma e moltiplicazione per scalare, forma esso stesso uno spazio vettoriale (notebook 6.3), di dimensione $n^{r+s}$ (notebook 6.9).

### Casi particolari già incontrati

- **Tipo $(1,0)$:** un'applicazione lineare $V^*\to\mathbb R$. Per la dualità tra $V$ e $V^{**}\cong V$ (un fatto di algebra lineare in dimensione finita), questo è esattamente un **vettore** $v\in V=T_pM$ (notebook 16.3): dato $v$, l'applicazione $\varphi\mapsto\varphi(v)$ è lineare in $\varphi$.
- **Tipo $(0,1)$:** un'applicazione lineare $V\to\mathbb R$, cioè esattamente un **covettore** $\omega\in V^*$ (notebook 16.4).
- **Tipo $(0,2)$:** un'applicazione bilineare $V\times V\to\mathbb R$. La **metrica** $g_p$ (notebook 16.6) è un tensore di tipo $(0,2)$, simmetrico e definito positivo — condizioni aggiuntive rispetto alla sola multilinearità, che selezionano un sottoinsieme speciale dei tensori $(0,2)$.
- **Tipo $(1,3)$:** il **tensore di curvatura di Riemann** $R$ (notebook 16.9), che prende un covettore e tre vettori (o, equivalentemente nella scrittura $R^\ell_{kij}$, ha un indice "alto" e tre "bassi") e restituisce un numero.
- **Tipo $(0,k)$, antisimmetrico:** una **$k$-forma differenziale** (notebook 16.4) è esattamente un tensore di tipo $(0,k)$ che soddisfa la condizione aggiuntiva di antisimmetria totale (scambiando due argomenti qualunque, il segno cambia) — le forme differenziali sono un caso particolare (antisimmetrico) dei tensori $(0,k)$, non tutti i tensori $(0,k)$ sono forme.

### Componenti e legge di trasformazione

Scelta una base $\{e_i\}$ di $V$ e la sua base duale $\{e^i\}$ (con $e^i(e_j)=\delta^i_j$, come nel notebook 16.4), un tensore $T$ di tipo $(r,s)$ ha **componenti** $T^{i_1\ldots i_r}_{j_1\ldots j_s}=T(e^{i_1},\ldots,e^{i_r},e_{j_1},\ldots,e_{j_s})$ — esattamente come una matrice (notebook 6.10, un tensore $(1,1)$) ha componenti $A^i_j$ rispetto a una base.

Se cambiamo base, $\tilde e_i=\sum_jP^j_ie_j$ per una matrice invertibile $P$ (notebook 6.16, matrice di cambiamento di base), le componenti di $T$ **si trasformano in modo prevedibile**:
$$\tilde T^{i_1\ldots i_r}_{j_1\ldots j_s}=\sum(P^{-1})^{i_1}_{k_1}\cdots(P^{-1})^{i_r}_{k_r}\,P^{\ell_1}_{j_1}\cdots P^{\ell_s}_{j_s}\,T^{k_1\ldots k_r}_{\ell_1\ldots\ell_s}$$
— un indice "alto" (in alto, controvariante) si trasforma con $P^{-1}$, come le componenti di un vettore rispetto a un cambio di base (notebook 6.10: le componenti di un vettore fisso cambiano con l'inversa della matrice che cambia i vettori di base, altrimenti il vettore stesso, combinazione lineare di base e componenti, non resterebbe invariato); un indice "basso" (covariante) si trasforma con $P$, come le componenti di un covettore.

**Questa legge di trasformazione è, di fatto, la definizione operativa di tensore usata più spesso in fisica e in molti testi**: una collezione di numeri, indicizzata da indici alti e bassi, è un tensore se e solo se si trasforma secondo questa regola cambiando coordinate. Le due definizioni (multilineare astratta, e "numeri che si trasformano così") sono equivalenti, e ciascuna ha i suoi vantaggi: la prima chiarisce il significato geometrico, la seconda rende immediato verificare o confutare che una quantità data sia un tensore.

### Un caso di studio: i simboli di Christoffel NON sono un tensore

Applichiamo il criterio della legge di trasformazione ai simboli di Christoffel $\Gamma^k_{ij}$ del notebook 16.8. Se fossero le componenti di un tensore di tipo $(1,2)$, dovrebbero trasformarsi omogeneamente come sopra. Ma un calcolo diretto (che omettiamo, essendo lungo, ma è uno dei fatti più noti della geometria differenziale) mostra che la legge di trasformazione di $\Gamma$ contiene un **termine aggiuntivo** che coinvolge le derivate seconde della funzione di transizione tra le coordinate — un termine che si annulla solo per cambi di coordinate **lineari**, ma non in generale. Questo spiega un fatto già osservato: in coordinate cartesiane euclidee $\Gamma\equiv0$ (notebook 16.8), ma in coordinate polari o sferiche (stesso spazio, $\mathbb R^n$ piatto) $\Gamma\ne0$ — se $\Gamma$ fosse un tensore, essendo nullo in una base dovrebbe essere nullo in **ogni** base (un tensore nullo lo è in ogni sistema di coordinate, per la linearità della legge di trasformazione, che manda $0$ in $0$): il fatto che questo non accada per $\Gamma$ conferma direttamente che non è un tensore. Il tensore di curvatura $R$ (notebook 16.9), costruito da $\Gamma$ e dalle sue derivate in una combinazione specifica, **è** invece un vero tensore: i termini "non tensoriali" delle derivate di $\Gamma$ si cancellano esattamente nella combinazione che definisce $R$.

```python
import sympy as sp

# Verifica diretta: i simboli di Christoffel dello spazio euclideo (piatto) sono
# nulli in coordinate cartesiane, ma NON nulli in coordinate polari -- se Gamma
# fosse un tensore, essendo zero in una base dovrebbe essere zero in ogni base.

r, phi = sp.symbols('r phi', positive=True)
# metrica euclidea in coordinate polari: ds^2 = dr^2 + r^2 dphi^2
g = sp.diag(1, r**2)
coords = [r, phi]
n = 2
g_inv = g.inv()

def christoffel(k, i, j):
    s = 0
    for l in range(n):
        s += g_inv[k, l] * (sp.diff(g[j, l], coords[i]) + sp.diff(g[i, l], coords[j]) - sp.diff(g[i, j], coords[l]))
    return sp.simplify(s / 2)

print("Simboli di Christoffel in coordinate polari (stesso spazio euclideo piatto!):")
for k in range(n):
    for i in range(n):
        for j in range(n):
            val = christoffel(k, i, j)
            if val != 0:
                nomi = ['r', 'phi']
                print(f"  Gamma^{nomi[k]}_{{{nomi[i]}{nomi[j]}}} = {val}")

print()
print("In coordinate cartesiane (x,y) gli stessi simboli sono TUTTI nulli (notebook 16.8).")
print("Se Gamma fosse un tensore, essendo zero in una base dovrebbe esserlo in ogni base:")
print("il fatto che compaiano termini non nulli qui conferma che Gamma NON e' un tensore.")
```

## Esercizio svolto e commentato

**Problema:** verifica che la curvatura di Gauss $K$ dello spazio euclideo piatto (calcolata usando i simboli di Christoffel in coordinate polari appena trovati) è comunque $0$, esattamente come in coordinate cartesiane — coerentemente col fatto che $K$, a differenza di $\Gamma$, **è** una quantità tensoriale (più precisamente, scalare, un tensore di tipo $(0,0)$), quindi il suo valore non può dipendere dalle coordinate scelte.

**Ragionamento:**

**Passo 1 — perché ci aspettiamo $K=0$ a prescindere dal calcolo.** Le coordinate polari descrivono lo **stesso** spazio euclideo piatto delle coordinate cartesiane, solo "visto" in modo diverso: un cambio di coordinate non può cambiare la geometria intrinseca dello spazio. Poiché $K$ è una quantità scalare intrinseca (un tensore di tipo $(0,0)$, invariante per cambio di coordinate essendo $r=s=0$ nella legge di trasformazione, che quindi non ha nessun indice da trasformare), il suo valore deve essere lo stesso $K=0$ trovato in coordinate cartesiane nel notebook 16.9 (esercizio 4).

**Passo 2 — calcoliamo comunque esplicitamente, per verifica.** Usiamo la formula del notebook 16.9, $R^\ell_{kij}=\partial_i\Gamma^\ell_{jk}-\partial_j\Gamma^\ell_{ik}+\Gamma^\ell_{im}\Gamma^m_{jk}-\Gamma^\ell_{jm}\Gamma^m_{ik}$, con i valori di $\Gamma$ appena calcolati (unici non nulli: $\Gamma^r_{\phi\phi}=-r$, $\Gamma^\phi_{r\phi}=\Gamma^\phi_{\phi r}=1/r$).

**Passo 3 — verifica con SymPy nella cella seguente.** Il calcolo, pur avendo $\Gamma$ non nullo, deve dare $R\equiv0$ e quindi $K=0$: la non-nullità di $\Gamma$ da sola **non implica** curvatura non nulla, proprio perché $\Gamma$ non è tensoriale e i suoi valori non nulli possono essere un semplice artefatto della scelta di coordinate curvilinee su uno spazio in realtà piatto — a differenza di $R$, che essendo tensoriale, se è nullo in coordinate polari lo è (ed era) anche in coordinate cartesiane, e viceversa.

```python
import sympy as sp

r, phi = sp.symbols('r phi', positive=True)
g = sp.diag(1, r**2)
coords = [r, phi]
n = 2
g_inv = g.inv()

def christoffel(k, i, j):
    s = 0
    for l in range(n):
        s += g_inv[k, l] * (sp.diff(g[j, l], coords[i]) + sp.diff(g[i, l], coords[j]) - sp.diff(g[i, j], coords[l]))
    return sp.simplify(s / 2)

Gamma = [[[christoffel(k, i, j) for j in range(n)] for i in range(n)] for k in range(n)]

def riemann(l, k, i, j):
    termine = sp.diff(Gamma[l][j][k], coords[i]) - sp.diff(Gamma[l][i][k], coords[j])
    for m in range(n):
        termine += Gamma[l][i][m]*Gamma[m][j][k] - Gamma[l][j][m]*Gamma[m][i][k]
    return sp.simplify(termine)

R_r_phi_r_phi = sum(g[0, l] * riemann(l, 1, 0, 1) for l in range(n))
det_g = sp.simplify(g[0,0]*g[1,1] - g[0,1]*g[1,0])
K = sp.simplify(sp.simplify(R_r_phi_r_phi) / det_g)

print("R_(r phi r phi) =", sp.simplify(R_r_phi_r_phi))
print("K in coordinate polari =", K, " (atteso: 0, come in coordinate cartesiane)")
```

**1.** Con $x=r\cos\phi,y=r\sin\phi$, la matrice Jacobiana è $P=\begin{pmatrix}\partial x/\partial r&\partial x/\partial\phi\\\partial y/\partial r&\partial y/\partial\phi\end{pmatrix}=\begin{pmatrix}\cos\phi&-r\sin\phi\\\sin\phi&r\cos\phi\end{pmatrix}$. **2.** Il gradiente $\nabla f$, come vettore in $T_p\mathbb R^n$, è di tipo $(1,0)$. L'operatore forma $S_p$, un'applicazione lineare $T_p\Sigma\to T_p\Sigma$, è di tipo $(1,1)$ (un indice alto per l'output vettoriale, uno basso per l'input, esattamente come una matrice quadrata generica, notebook 6.10, letta come applicazione lineare).

**3.** Se $T,S$ sono entrambi multilineari di tipo $(r,s)$, la loro somma $(T+S)(\varphi_1,\ldots,\varphi_r,v_1,\ldots,v_s):=T(\ldots)+S(\ldots)$ è ancora multilineare: fissati tutti gli argomenti tranne uno, linearità in quell'argomento singolo segue immediatamente dalla linearità di $T$ e di $S$ separatamente in quell'argomento, sommando le due identità di linearità. Quindi $T+S$ è ancora un tensore di tipo $(r,s)$: insieme alla moltiplicazione per scalare (verifica analoga), questo conferma che i tensori $(r,s)$ soddisfano gli assiomi di spazio vettoriale (notebook 6.3).

**4.** $(dx\otimes dx+dy\otimes dy)(v,w)=dx(v)\,dx(w)+dy(v)\,dy(w)$ (per definizione di prodotto tensoriale e di somma di tensori). $dx(v)=v_1$ (prima componente di $v$, per definizione della base duale, notebook 16.4), $dx(w)=w_1$, $dy(v)=v_2$, $dy(w)=w_2$. ## Riepilogo finale

- Un **tensore di tipo $(r,s)$** è un'applicazione multilineare su $r$ covettori e $s$ vettori; vettori, covettori, la metrica, e il tensore di curvatura sono tutti casi particolari, di tipo $(1,0)$, $(0,1)$, $(0,2)$ e $(1,3)$ rispettivamente.
- Le componenti di un tensore si trasformano secondo una legge precisa cambiando base: indici alti con $P^{-1}$, indici bassi con $P$ — questa legge è spesso usata come definizione operativa alternativa.
- I simboli di Christoffel $\Gamma^k_{ij}$ **non** sono le componenti di un tensore (possono essere nulli in una base e non nulli in un'altra, come mostrato passando da coordinate cartesiane a polari sullo stesso spazio piatto), mentre il tensore di curvatura $R$, costruito da $\Gamma$ in una combinazione specifica, **è** un vero tensore.
- Il **prodotto tensoriale** $\otimes$ combina tensori di tipo diverso in un tensore di tipo più alto, fornendo, ad esempio, un modo di scrivere la metrica euclidea come $\sum_idx^i\otimes dx^i$.
- Questo linguaggio unificato — usato per la prima volta esplicitamente in questo notebook, ma implicitamente presente in ogni notebook precedente del modulo — è quello standard della geometria differenziale moderna e della relatività generale.
