# 16.3 — Spazi tangenti

**Prerequisiti:** 16.2 Superfici, 6.3 Spazi vettoriali, 7.7 Derivate parziali, 5.14 Formula di Taylor.


## Perché ci interessa

Nel notebook 16.2, il piano tangente a una superficie è stato costruito usando i vettori $\sigma_u,\sigma_v$, che vivono dentro $\mathbb R^3$: lo spazio ambiente in cui la superficie è immersa. Ma questa costruzione ha un difetto: dipende dal fatto che la superficie **sia dentro** $\mathbb R^3$. Nel notebook 16.5 vogliamo studiare spazi curvi — varietà — che non sono definiti come sottoinsiemi di nessuno spazio ambiente più grande: sono lo spazio stesso, non un oggetto dentro un altro spazio.

Per farlo servono i vettori tangenti, ma **senza spazio ambiente in cui immergerli**. Questo notebook risolve il problema: dà una definizione di vettore tangente che usa solo cose intrinseche al punto stesso (le funzioni definite vicino a quel punto), utilizzabile identicamente in $\mathbb R^n$ e, dopo il notebook 16.5, su una varietà astratta qualunque.

## Teoria, passo dopo passo

### Il problema con la definizione "estrinseca"

Nel notebook 16.2, un vettore tangente era un vettore di $\mathbb R^3$, ottenuto come combinazione lineare di $\sigma_u,\sigma_v$. Ma cos'è, davvero, un vettore tangente, indipendentemente dal fatto che lo pensiamo "dentro" $\mathbb R^3$? Due idee equivalenti rispondono a questa domanda.

### Prima idea: vettori tangenti come classi di curve

Fissiamo un punto $p\in\mathbb R^n$ (per ora lavoriamo ancora in $\mathbb R^n$, non su una superficie astratta: la generalizzazione arriverà nel notebook 16.5). Consideriamo tutte le curve $\gamma:(-\varepsilon,\varepsilon)\to\mathbb R^n$ (notebook 16.1) con $\gamma(0)=p$. Diciamo che due tali curve $\gamma_1,\gamma_2$ sono **equivalenti in $p$** se $\gamma_1'(0)=\gamma_2'(0)$: stessa velocità iniziale. Un **vettore tangente** in $p$ è allora una classe di equivalenza $[\gamma]$ di curve.

Questa definizione è puramente intrinseca: parla solo di curve che passano per $p$, senza mai fare riferimento a un piano tangente pre-esistente dentro uno spazio ambiente. In $\mathbb R^n$, ogni classe $[\gamma]$ corrisponde esattamente al vettore $\gamma'(0)\in\mathbb R^n$: la definizione, qui, non aggiunge nulla di nuovo, ma il punto è che **si può ripetere identica su qualunque spazio dove abbia senso parlare di curve e di derivarle**, anche uno che non sia un sottoinsieme di $\mathbb R^n$.

### Seconda idea: vettori tangenti come derivazioni

Un'idea equivalente, più algebrica: un vettore tangente $v$ in $p$ agisce su ogni funzione derivabile $f$ definita vicino a $p$, restituendo la **derivata direzionale** $v(f)=\dfrac{d}{dt}\Big|_{t=0}f(\gamma(t))$, dove $\gamma$ è una curva con $\gamma(0)=p,\gamma'(0)=v$ (per la regola della catena, notebook 7.20, questo numero non dipende dalla scelta di $\gamma$ nella sua classe di equivalenza, solo da $v=\gamma'(0)$). Questa operazione $f\mapsto v(f)$ soddisfa due proprietà, verificabili direttamente dalle regole di derivazione (notebook 5.9 e 7.20):

- **Linearità:** $v(af+bg)=a\,v(f)+b\,v(g)$;
- **Regola di Leibniz:** $v(fg)=v(f)\,g(p)+f(p)\,v(g)$.

Un'applicazione lineare sullo spazio delle funzioni derivabili vicino a $p$ che soddisfa la regola di Leibniz si chiama **derivazione** in $p$. Si dimostra (usando la formula di Taylor, notebook 5.14, applicata a $f$ vicino a $p$) che **ogni** derivazione in $p$ è della forma $f\mapsto v(f)$ per un unico vettore $v\in\mathbb R^n$: le derivazioni in $p$ e i vettori tangenti (nel senso della prima idea) sono esattamente la stessa cosa, vista da due angolazioni diverse. Questa seconda descrizione è quella che si generalizza più facilmente a varietà astratte, perché non richiede nemmeno di poter "vedere" le curve dall'esterno.

### Lo spazio tangente

L'insieme di tutti i vettori tangenti in $p$, con le operazioni naturali $[\gamma_1]+[\gamma_2]$ definita tramite $(\gamma_1'(0)+\gamma_2'(0))$ e $\lambda[\gamma]$ tramite $\lambda\gamma'(0)$, forma uno **spazio vettoriale** $T_p\mathbb R^n$ (notebook 6.3), chiamato **spazio tangente** in $p$. In $\mathbb R^n$, $T_p\mathbb R^n$ è (canonicamente) isomorfo a $\mathbb R^n$ stesso: ha dimensione $n$ (notebook 6.9), con base naturale $\Big\{\dfrac{\partial}{\partial x_1}\Big|_p,\ldots,\dfrac{\partial}{\partial x_n}\Big|_p\Big\}$, dove $\dfrac{\partial}{\partial x_i}\Big|_p$ è la derivazione "deriva parzialmente rispetto a $x_i$, poi valuta in $p$" — corrispondente, nella prima descrizione, alla curva $\gamma(t)=p+te_i$ (con $e_i$ $i$-esimo vettore della base canonica, notebook 6.1).

**Perché la distinzione tra $\mathbb R^n$ e $T_p\mathbb R^n$, se sono isomorfi?** Perché tenerli concettualmente separati prepara il terreno per il caso in cui non sono più isomorfi in modo canonico: su una superficie curva (e ancora di più su una varietà astratta), lo spazio tangente in punti diversi $p\ne q$ **non ha alcuna identificazione naturale** con $T_q$ — sono spazi vettoriali diversi, ciascuno legato al proprio punto, senza un modo canonico di confrontare un vettore tangente in $p$ con uno in $q$. Il notebook 16.8 (connessioni) introdurrà uno strumento — il trasporto parallelo — proprio per confrontare spazi tangenti in punti diversi lungo una curva.

### Riconciliazione col piano tangente del notebook 16.2

Per una superficie regolare $\Sigma=\sigma(U)\subset\mathbb R^3$ e un punto $p=\sigma(u_0,v_0)$, lo spazio tangente $T_p\Sigma$ (definito come lo spazio delle velocità $\gamma'(0)$ di curve $\gamma$ **che restano su $\Sigma$**, con $\gamma(0)=p$) coincide esattamente con il piano generato da $\sigma_u,\sigma_v$ del notebook 16.2: infatti ogni curva $\gamma(t)=\sigma(u(t),v(t))$ sulla superficie ha, per la regola della catena, $\gamma'(0)=u'(0)\sigma_u+v'(0)\sigma_v$, un generico elemento di quel piano; viceversa ogni tale combinazione lineare si ottiene scegliendo opportunamente $u(t),v(t)$. La definizione astratta di questo notebook e quella concreta del notebook 16.2 descrivono, per le superfici, esattamente lo stesso oggetto — solo che ora sappiamo formularlo senza fare riferimento a $\mathbb R^3$.

```python
import sympy as sp

# Verifica: la derivazione "direzionale" lungo una curva non dipende
# dalla scelta della curva nella classe di equivalenza, solo da gamma'(0).
t, x, y = sp.symbols('t x y', real=True)
f = x**2 * y + sp.sin(x*y)   # una funzione f(x,y) qualunque

# Curva 1: gamma1(t) = (1+t, 2+3t)  --> gamma1'(0) = (1,3)
gamma1 = (1 + t, 2 + 3*t)
# Curva 2: gamma2(t) = (1+t+t**2, 2+3t+5t**2) --> gamma2'(0) = (1,3) anche qui!
gamma2 = (1 + t + t**2, 2 + 3*t + 5*t**2)

f1 = f.subs({x: gamma1[0], y: gamma1[1]})
f2 = f.subs({x: gamma2[0], y: gamma2[1]})

d1 = sp.diff(f1, t).subs(t, 0)
d2 = sp.diff(f2, t).subs(t, 0)

print("d/dt f(gamma1(t)) in t=0:", sp.simplify(d1))
print("d/dt f(gamma2(t)) in t=0:", sp.simplify(d2))
print("Coincidono anche se le curve sono diverse (stessa velocita' iniziale (1,3)):", sp.simplify(d1 - d2) == 0)

# Confronto con v(f) = grad(f) . v, v=(1,3), calcolato in p=(1,2)
grad_f = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
v = sp.Matrix([1, 3])
vf = grad_f.dot(v).subs({x: 1, y: 2})
print("Formula v(f) = grad(f)(p) . v:", sp.simplify(vf))
```

## Esercizio svolto e commentato

**Problema:** in $\mathbb R^2$, nel punto $p=(1,1)$, considera le due curve $\gamma_1(t)=(1+t,e^t)$ e $\gamma_2(t)=(\cos t+ \sin t, 1+t+t^3)$. Verifica che sono equivalenti in $p$ (stessa classe $[\gamma]$) e usa questo fatto per calcolare $v(f)$ per $f(x,y)=x^3y$ **senza dover derivare $f$ lungo entrambe le curve**.

**Ragionamento:**

**Passo 1 — verifichiamo $\gamma_1(0)=\gamma_2(0)=p$.** $\gamma_1(0)=(1,e^0)=(1,1)=p$ ✓. $\gamma_2(0)=(\cos0+\sin0,1+0+0)=(1,1)=p$ ✓.

**Passo 2 — calcoliamo le velocità iniziali.** $\gamma_1'(t)=(1,e^t)\Rightarrow\gamma_1'(0)=(1,1)$. $\gamma_2'(t)=(-\sin t+\cos t,1+3t^2)\Rightarrow\gamma_2'(0)=(0+1,1+0)=(1,1)$.

**Passo 3 — concludiamo che sono equivalenti.** $\gamma_1'(0)=\gamma_2'(0)=(1,1)=:v$: per definizione, $\gamma_1\sim\gamma_2$, cioè rappresentano lo **stesso** vettore tangente in $p$.

**Passo 4 — perché questo ci fa risparmiare lavoro.** Per la teoria (derivazioni), $v(f)$ dipende solo dalla classe $[\gamma]$, cioè solo da $v=(1,1)$, non da quale rappresentante scegliamo. Quindi possiamo calcolare $v(f)$ nel modo più comodo: $v(f)=\nabla f(p)\cdot v$ (notebook 7.9, gradiente). $\nabla f=(3x^2y,x^3)$, quindi $\nabla f(1,1)=(3,1)$, e $v(f)=(3,1)\cdot(1,1)=3+1=4$.

**Passo 5 — verifica che dà lo stesso risultato lungo entrambe le curve.** Se avessimo calcolato $\frac{d}{dt}\big|_{t=0}f(\gamma_1(t))$ e $\frac{d}{dt}\big|_{t=0}f(\gamma_2(t))$ direttamente (un calcolo più lungo, specialmente per $\gamma_2$), avremmo dovuto ottenere lo stesso valore $4$ — verifichiamolo con SymPy nella cella seguente, come conferma della teoria.

```python
import sympy as sp

t, x, y = sp.symbols('t x y', real=True)
f = x**3 * y

gamma1 = (1 + t, sp.exp(t))
gamma2 = (sp.cos(t) + sp.sin(t), 1 + t + t**3)

for nome, gamma in [("gamma1", gamma1), ("gamma2", gamma2)]:
    f_su_gamma = f.subs({x: gamma[0], y: gamma[1]})
    derivata = sp.diff(f_su_gamma, t).subs(t, 0)
    print(f"d/dt f({nome}(t)) in t=0:", sp.simplify(derivata))

grad_f = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
v = sp.Matrix([1, 1])
vf = grad_f.dot(v).subs({x: 1, y: 1})
print("Formula v(f) = grad(f)(p).v:", vf)
```

## Riepilogo finale

- Un vettore tangente in un punto $p$ può essere definito in modo intrinseco, senza spazio ambiente, come classe di equivalenza di curve passanti per $p$ con la stessa velocità iniziale, o equivalentemente come derivazione (operatore lineare che soddisfa la regola di Leibniz).
- Le due descrizioni coincidono: ogni derivazione è della forma $f\mapsto\frac{d}{dt}\big|_0f(\gamma(t))$ per un unico $v$.
- L'insieme dei vettori tangenti in $p$ forma uno spazio vettoriale $T_p$, di dimensione uguale a quella dello spazio ambiente (per un punto in $\mathbb R^n$) o della superficie stessa (per un punto su una superficie regolare).
- Spazi tangenti in punti diversi non hanno un'identificazione canonica: questa osservazione, apparentemente astratta, motiverà la necessità di una connessione (notebook 16.8) per confrontare vettori tangenti lungo una curva.
- Su una superficie regolare, questa definizione astratta coincide esattamente con il piano tangente concreto del notebook 16.2, costruito con $\sigma_u,\sigma_v$.
