## Perché ci interessa

Nei notebook 16.1-16.4 abbiamo sempre lavorato con oggetti **immersi**: curve e superfici dentro $\mathbb R^n$, spazi tangenti pensati come sottospazi di $\mathbb R^n$. Ma la superficie terrestre, per chi ci vive sopra, non è "immersa" in nulla: è semplicemente lo spazio in cui si vive, e localmente sembra un piano, anche se globalmente è una sfera. Questo notebook formalizza questa idea: una **varietà** è uno spazio che localmente assomiglia a $\mathbb R^n$, senza bisogno di alcuno spazio ambiente più grande in cui immergerlo.

La definizione userà pesantemente la topologia generale (Modulo 12): serviranno spazi topologici, omeomorfismi (per dire cosa significa "assomigliare localmente" a $\mathbb R^n$), e spazi prodotto/quoziente (per costruire esempi importanti come il toro). Una volta definita la varietà, gli spazi tangenti e le forme differenziali dei notebook 16.3-16.4 si trasportano su di essa quasi senza modifiche: avevamo già fatto il lavoro concettuale più difficile.

## Teoria, passo dopo passo

### Carte locali

Sia $M$ uno spazio topologico (notebook 12.1) di Hausdorff e a base numerabile (ipotesi tecniche che escludono patologie, come punti "doppi" indistinguibili topologicamente). Una **carta** su $M$ è una coppia $(U,\varphi)$, dove $U\subseteq M$ è aperto (notebook 12.3) e $\varphi:U\to\varphi(U)\subseteq\mathbb R^n$ è un **omeomorfismo** (notebook 12.11) tra $U$ e un aperto di $\mathbb R^n$. La carta fornisce delle "coordinate locali" su $U$: ogni punto $p\in U$ è identificato dalla $n$-upla $\varphi(p)=(x_1(p),\ldots,x_n(p))\in\mathbb R^n$.

### Atlanti e varietà topologiche

Un **atlante** è una collezione di carte $\{(U_\alpha,\varphi_\alpha)\}$ che ricopre $M$, cioè $\bigcup_\alpha U_\alpha=M$. $M$, munito di un atlante, si chiama **varietà topologica** di dimensione $n$: uno spazio in cui, vicino a ogni punto, "sembra" di stare in $\mathbb R^n$, anche se globalmente $M$ può avere una forma completamente diversa (come una sfera o un toro).

### Compatibilità delle carte e varietà differenziabili

Se due carte $(U_\alpha,\varphi_\alpha)$ e $(U_\beta,\varphi_\beta)$ hanno intersezione $U_\alpha\cap U_\beta\ne\emptyset$, la **funzione di transizione** $\varphi_\beta\circ\varphi_\alpha^{-1}:\varphi_\alpha(U_\alpha\cap U_\beta)\to\varphi_\beta(U_\alpha\cap U_\beta)$ è una funzione tra aperti di $\mathbb R^n$ (composizione di omeomorfismi, quindi già un omeomorfismo, notebook 12.11). Se, in più, ogni funzione di transizione è $C^\infty$ (derivabile infinite volte, come funzione tra aperti di $\mathbb R^n$, nel senso del Modulo 7), l'atlante si dice **liscio**, e $M$ con questo atlante è una **varietà differenziabile**. Questa condizione di compatibilità è ciò che permette di parlare in modo coerente di funzioni derivabili, vettori tangenti e forme differenziali su $M$: definiti in una carta, la compatibilità liscia garantisce che le nozioni non dipendano dalla carta scelta.

### Esempi

- **$\mathbb R^n$ stesso**, con l'unica carta $(\mathbb R^n,\mathrm{id})$: l'esempio banale.
- **La sfera $S^n=\{x\in\mathbb R^{n+1}:\|x\|=1\}$**, con topologia di sottospazio (notebook 12.1, restrizione della topologia di $\mathbb R^{n+1}$): un atlante minimale usa due carte, le **proiezioni stereografiche** dal polo nord e dal polo sud, ciascuna un omeomorfismo tra $S^n$ meno un punto e $\mathbb R^n$ — nessuna singola carta può coprire tutta $S^n$ (che è compatta, notebook 12.6, mentre $\mathbb R^n$ non lo è, quindi non possono essere omeomorfe per intero).
- **Il toro $T^2=S^1\times S^1$**: uno **spazio prodotto** (notebook 12.9) di due circonferenze. Un atlante si costruisce a partire dagli atlanti delle due circonferenze fattore.
- **Il toro come quoziente**, $T^2=\mathbb R^2/\mathbb Z^2$: uno **spazio quoziente** (notebook 12.10), identificando $(x,y)\sim(x+m,y+k)$ per $m,k\in\mathbb Z$. Le carte si ottengono restringendo la proiezione al quoziente a piccoli quadrati di $\mathbb R^2$, dove la proiezione è iniettiva e quindi (essendo anche continua e aperta) un omeomorfismo sull'immagine. Questa costruzione mostra concretamente **la stessa varietà descritta in due modi diversi** (prodotto ed equivalentemente quoziente), un fatto non banale che si dimostra costruendo esplicitamente un omeomorfismo tra le due presentazioni.
- **Lo spazio proiettivo reale $\mathbb{RP}^n$**: lo spazio quoziente di $\mathbb R^{n+1}\setminus\{0\}$ (o, equivalentemente, di $S^n$) rispetto alla relazione $x\sim\lambda x$ per $\lambda\ne0$ (notebook 12.10): l'insieme delle rette per l'origine in $\mathbb R^{n+1}$.

### Spazi tangenti e forme differenziali su una varietà

Tutto quello che abbiamo costruito nei notebook 16.3-16.4 si trasporta: in un punto $p\in M$, scelta una carta $(U,\varphi)$ intorno a $p$, lo spazio tangente $T_pM$ è definito esattamente come nel notebook 16.3, usando funzioni $f\circ\varphi^{-1}$ (che ora sono funzioni ordinarie su un aperto di $\mathbb R^n$) al posto di $f$. La condizione di compatibilità liscia delle carte garantisce che, cambiando carta, lo spazio tangente ottenuto sia "lo stesso" (isomorfo tramite una trasformazione lineare esplicita, la matrice Jacobiana della funzione di transizione, notebook 7.20). Analogamente, forme differenziali e la derivata esterna $d$ (notebook 16.4) si definiscono carta per carta, e la compatibilità liscia assicura che siano ben definite su tutto $M$.

**Il punto essenziale:** una volta stabilita questa compatibilità, **tutta la geometria differenziale che seguirà (metriche, geodetiche, curvatura, notebook 16.6-16.9) si sviluppa su $M$ esattamente come se fossimo in $\mathbb R^n$**, lavorando in coordinate locali una carta alla volta, con la garanzia che i risultati non dipendano dalla carta scelta.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Visualizziamo le due carte della sfera S^2 tramite proiezione stereografica
# dal polo nord (0,0,1): manda (x,y,z) in (x/(1-z), y/(1-z))
theta = np.linspace(0.05, np.pi - 0.05, 40)
phi = np.linspace(0, 2*np.pi, 40)
Theta, Phi = np.meshgrid(theta, phi)
X = np.sin(Theta)*np.cos(Phi)
Y = np.sin(Theta)*np.sin(Phi)
Z = np.cos(Theta)

fig = plt.figure(figsize=(12, 5))

ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X, Y, Z, alpha=0.6, cmap='viridis')
ax1.set_title("La sfera $S^2$ (varieta' astratta)")

# proiezione stereografica dal polo nord: definita ovunque tranne il polo nord stesso
U = X / (1 - Z)
V = Y / (1 - Z)
ax2 = fig.add_subplot(122)
ax2.scatter(U.flatten(), V.flatten(), s=2)
ax2.set_title("Carta: proiezione stereografica dal polo nord\n(omeomorfismo con un aperto di $R^2$)")
ax2.set_aspect('equal')
plt.tight_layout()
plt.show()
print("Nessuna singola carta copre tutta la sfera: il polo nord (Z=1) manda a infinito.")
print("Servono almeno due carte (proiezione da nord e da sud) per un atlante completo.")
```

## Esercizio svolto e commentato

**Problema:** verifica esplicitamente che la funzione di transizione tra le due carte del toro $T^2=\mathbb R^2/\mathbb Z^2$ (proiezione ristretta a due quadrati che si sovrappongono) è liscia, in un caso concreto.

**Ragionamento:**

**Passo 1 — fissiamo due carte concrete.** Sia $\pi:\mathbb R^2\to T^2$ la proiezione al quoziente (notebook 12.10). Prendiamo $U_1=\pi\big((-0.3,0.6)\times(-0.3,0.6)\big)$ con carta $\varphi_1=\pi^{-1}$ ristretta a quel quadrato (cioè $\varphi_1$ manda il punto del toro nel rappresentante che sta in quel quadrato), e $U_2=\pi\big((0.2,1.1)\times(0.2,1.1)\big)$ con carta analoga $\varphi_2$.

**Passo 2 — troviamo la regione di sovrapposizione.** $U_1\cap U_2$ corrisponde, nella carta $\varphi_1$, ai punti $(x,y)\in(0.2,0.6)\times(0.2,0.6)$ (l'intersezione dei due quadrati, dato che qui i rappresentanti coincidono senza bisogno di traslare per interi).

**Passo 3 — scriviamo la funzione di transizione su questa regione.** Per $(x,y)\in(0.2,0.6)^2$, il punto $\pi(x,y)$ ha, nella carta $\varphi_2$, lo stesso rappresentante $(x,y)$ (perché $(x,y)$ appartiene già anche al secondo quadrato $(0.2,1.1)^2$, quindi non serve sommare interi). Quindi $\varphi_2\circ\varphi_1^{-1}(x,y)=(x,y)$ su questa sotto-regione: **la funzione identità**, banalmente $C^\infty$.

**Passo 4 — cosa succede su una sovrapposizione "al bordo"?** Se avessimo scelto $U_2=\pi\big((0.7,1.6)\times(-0.3,0.6)\big)$, la sovrapposizione con $U_1$ nella prima coordinata sarebbe $x\in(0.7,0.6)$... vuota in quell'ordine, quindi serve guardare vicino a $x=1\equiv0$: un punto con rappresentante $x=0.75$ nella carta $\varphi_1$ (se $U_1$ include quella zona modulo l'identificazione) avrebbe rappresentante $x=0.75$ anche in $\varphi_2$ senza bisogno di traslare, oppure, per punti con $\varphi_1$-rappresentante vicino a $x=-0.25$ (cioè vicino a $1$ nel verso opposto), la transizione dovrebbe sommare $+1$: $\varphi_2\circ\varphi_1^{-1}(x,y)=(x+1,y)$ su quella porzione. **Anche questa è $C^\infty$** (una traslazione è una funzione liscia, con Jacobiano l'identità).

**Conclusione:** in ogni caso la funzione di transizione è, localmente, o l'identità o una traslazione per un vettore intero — entrambe funzioni $C^\infty$ con inversa $C^\infty$: l'atlante del toro costruito come quoziente $\mathbb R^2/\mathbb Z^2$ è effettivamente liscio, confermando che questa presentazione definisce una legittima varietà differenziabile.

```python
import numpy as np

def rappresentante_in_quadrato(punto, angolo_basso_sinistro, lato=1.0):
    # riporta 'punto' (in R^2) al rappresentante equivalente (mod Z^2)
    # che cade nel quadrato [ax, ax+lato) x [ay, ay+lato)
    ax, ay = angolo_basso_sinistro
    x, y = punto
    x_r = ax + (x - ax) % lato
    y_r = ay + (y - ay) % lato
    return (x_r, y_r)

# Verifica: un punto vicino al "bordo" del toro, transizione tra due carte
punto_astratto = (0.9, 0.4)  # un punto sul toro, descritto da un rappresentante qualunque

rappr1 = rappresentante_in_quadrato(punto_astratto, (-0.3, -0.3))
rappr2 = rappresentante_in_quadrato(punto_astratto, (0.7, -0.3))

print("Rappresentante nella carta 1 (quadrato base [-0.3, 0.7)^2):", rappr1)
print("Rappresentante nella carta 2 (quadrato base [0.7, 1.7)^2):", rappr2)
print("Differenza (deve essere un vettore intero, es. (1,0) o (0,0)):",
      (rappr2[0]-rappr1[0], rappr2[1]-rappr1[1]))
```

**1.** $S^n$ (per $n\ge1$) è chiusa e limitata in $\mathbb R^{n+1}$, quindi compatta (teorema di Heine-Borel, richiamato nel notebook 12.6). $\mathbb R^n$ non è compatto (non è limitato). Quindi $f$ è un omeomorfismo.

**4.** Sia $J$ la matrice Jacobiana di $\varphi_2\circ\varphi_1^{-1}$ in $\varphi_1(p)$. Poiché $\varphi_2\circ\varphi_1^{-1}$ ha inversa liscia $\varphi_1\circ\varphi_2^{-1}$, per la regola della catena (notebook 7.20) applicata a $(\varphi_1\circ\varphi_2^{-1})\circ(\varphi_2\circ\varphi_1^{-1})=\mathrm{id}$, il prodotto delle due Jacobiane è la matrice identità: $J$ è invertibile (notebook 6.16), con inversa la Jacobiana della transizione opposta. ## Riepilogo finale

- Una **varietà topologica** di dimensione $n$ è uno spazio topologico ricoperto da carte, ciascuna un omeomorfismo con un aperto di $\mathbb R^n$; se le funzioni di transizione tra carte sono $C^\infty$, si parla di **varietà differenziabile**.
- Esempi fondamentali (sfera, toro, spazio proiettivo) mostrano come oggetti globalmente diversi da $\mathbb R^n$ possano comunque "assomigliargli" localmente, usando gli strumenti del Modulo 12 (spazi prodotto, quoziente, omeomorfismi).
- Spazi tangenti e forme differenziali (notebook 16.3-16.4), definiti carta per carta, sono ben definiti su tutta la varietà grazie alla compatibilità liscia delle carte: la matrice Jacobiana della funzione di transizione fornisce l'isomorfismo canonico tra le descrizioni in carte diverse.
- Con questa struttura in mano, tutta la geometria differenziale successiva (metriche, geodetiche, curvatura) si sviluppa esattamente come in $\mathbb R^n$, ma **una carta alla volta**, con la garanzia che i risultati siano indipendenti dalla scelta della carta.
