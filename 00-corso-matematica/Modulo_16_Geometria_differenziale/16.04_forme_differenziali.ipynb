# 16.4 — Forme differenziali

**Prerequisiti:** 16.3 Spazi tangenti, 6.3-6.9 Spazi vettoriali, basi e dimensione, 7.9 Gradiente, 7.20-7.22 Campi vettoriali, divergenza e rotore.


## Perché ci interessa

Nel Modulo 7 abbiamo incontrato tre operazioni apparentemente diverse su funzioni e campi vettoriali: il **gradiente** (che trasforma una funzione scalare in un campo vettoriale, notebook 7.9), la **divergenza** (che trasforma un campo vettoriale in una funzione scalare, notebook 7.21) e il **rotore** (che trasforma un campo vettoriale in un altro campo vettoriale, in $\mathbb R^3$, notebook 7.22). Sembrano tre strumenti indipendenti, ciascuno con la propria formula da ricordare.

Questo notebook mostra che sono **la stessa operazione**, applicata a oggetti di grado diverso: le **forme differenziali**. Il vantaggio non è solo estetico: le forme differenziali funzionano in qualunque dimensione (non solo $\mathbb R^3$) e, cosa più importante per il seguito del modulo, si possono definire su una varietà astratta qualunque (notebook 16.5), dove gradiente/divergenza/rotore separatamente non avrebbero nemmeno senso. Le forme differenziali useranno lo spazio tangente costruito nel notebook 16.3 — anzi, il suo **duale**.

## Teoria, passo dopo passo

### Covettori: lo spazio duale

Dato uno spazio vettoriale $T_p$ (lo spazio tangente in $p$, notebook 16.3), il suo **spazio duale** $T_p^*$ è l'insieme delle applicazioni lineari $\varphi:T_p\to\mathbb R$ (notebook 6.13, applicazioni lineari, qui a valori in $\mathbb R^1$). Ogni tale $\varphi$ si chiama **covettore**. Come $T_p$, anche $T_p^*$ è uno spazio vettoriale (con somma e moltiplicazione per scalare definite punto per punto sui valori) della stessa dimensione $n$ di $T_p$ (notebook 6.9): una base di $T_p$, per esempio $\Big\{\dfrac{\partial}{\partial x_i}\Big|_p\Big\}$, ha una **base duale** naturale $\{dx_1,\ldots,dx_n\}$ definita da $dx_i\Big(\dfrac{\partial}{\partial x_j}\Big|_p\Big)=\delta_{ij}$ (delta di Kronecker, come nel notebook 16.3, esercizio 3).

### Il differenziale di una funzione

Per una funzione derivabile $f$, il **differenziale** $df_p$ è il covettore in $p$ definito da $df_p(v)=v(f)$ — cioè applicare $df_p$ a un vettore tangente $v$ restituisce la derivata direzionale di $f$ lungo $v$ (notebook 16.3). Nella base duale, $df_p=\displaystyle\sum_{i=1}^n\frac{\partial f}{\partial x_i}(p)\,dx_i$: le componenti di $df_p$ sono esattamente le derivate parziali di $f$, cioè le componenti del gradiente $\nabla f(p)$ (notebook 7.9). **Il differenziale $df$ e il gradiente $\nabla f$ hanno le stesse componenti**, ma sono concettualmente diversi: $\nabla f$ è un vettore (vive in $T_p$), $df$ è un covettore (vive in $T_p^*$) — la distinzione, invisibile finché lavoriamo solo in $\mathbb R^n$ con coordinate cartesiane, diventerà essenziale sulle varietà (notebook 16.5), dove non c'è un prodotto scalare canonico che permetta di "confondere" $T_p$ con $T_p^*$ (il notebook 16.6 introdurrà la metrica proprio per costruire artificialmente questa identificazione).

### $k$-forme differenziali

Una **1-forma** su un aperto $U$ è una scelta liscia di covettore $\omega_p\in T_p^*$ per ogni $p\in U$; nella base $\{dx_i\}$ si scrive $\omega=\sum_ia_i(x)\,dx_i$, con $a_i$ funzioni lisce. Il differenziale $df$ è l'esempio più semplice di 1-forma.

Per andare oltre le 1-forme serve il **prodotto esterno** (wedge) $\wedge$, un'operazione antisimmetrica tra covettori: $dx_i\wedge dx_j=-dx_j\wedge dx_i$ (in particolare $dx_i\wedge dx_i=0$). Una **$k$-forma** è una combinazione lineare di prodotti $dx_{i_1}\wedge\cdots\wedge dx_{i_k}$ con coefficienti funzioni lisce. Ad esempio, in $\mathbb R^3$, una 2-forma generica è $\omega=a\,dx\wedge dy+b\,dy\wedge dz+c\,dz\wedge dx$, e l'unica 3-forma (a meno di riscalare) è $dx\wedge dy\wedge dz$ (ogni 4-forma in $\mathbb R^3$ è automaticamente zero, per antisimmetria: con solo 3 indici disponibili, un prodotto di 4 fattori $dx_i$ deve ripeterne almeno uno).

### La derivata esterna $d$

L'operazione $d$ generalizza il differenziale a forme di ogni grado: se $\omega=\sum a_I\,dx_I$ (con $dx_I=dx_{i_1}\wedge\cdots\wedge dx_{i_k}$), si definisce $d\omega=\sum da_I\wedge dx_I$, usando il differenziale di funzione già visto per ogni coefficiente $a_I$. La proprietà cruciale, verificabile direttamente dalla definizione (usando l'uguaglianza delle derivate parziali miste, un fatto standard per funzioni sufficientemente lisce), è:
$$d(d\omega)=0\quad\text{sempre.}$$

### Gradiente, rotore, divergenza come casi particolari di $d$

In $\mathbb R^3$, con coordinate $(x,y,z)$:

- **$d$ su una 0-forma (funzione) $f$:** $df=f_x\,dx+f_y\,dy+f_z\,dz$ — le componenti sono esattamente $\nabla f$. **$d$ sulle funzioni è il gradiente.**
- **$d$ su una 1-forma $\omega=P\,dx+Q\,dy+R\,dz$** (pensata come il campo vettoriale $F=(P,Q,R)$, notebook 7.20): un calcolo diretto usando $d(P\,dx)=dP\wedge dx=(P_y\,dy+P_z\,dz)\wedge dx=P_y\,dy\wedge dx+P_z\,dz\wedge dx=-P_y\,dx\wedge dy+P_z\,dz\wedge dx$ (usando l'antisimmetria) e termini analoghi per $Q\,dy,R\,dz$, dà $d\omega=(R_y-Q_z)\,dy\wedge dz+(P_z-R_x)\,dz\wedge dx+(Q_x-P_y)\,dx\wedge dy$: le componenti sono esattamente quelle del **rotore** $\nabla\times F$ (notebook 7.22). **$d$ sui campi vettoriali (1-forme) è il rotore.**
- **$d$ su una 2-forma $\eta=P\,dy\wedge dz+Q\,dz\wedge dx+R\,dx\wedge dy$** (di nuovo associata a $F=(P,Q,R)$, ma stavolta come "flusso"): un calcolo analogo dà $d\eta=(P_x+Q_y+R_z)\,dx\wedge dy\wedge dz$ — il coefficiente è esattamente la **divergenza** $\nabla\cdot F$ (notebook 7.21). **$d$ sulle 2-forme è la divergenza.**

### $d^2=0$ ricostruisce due identità del Modulo 7

Applicando $d^2=0$ a una funzione $f$ (0-forma): $d(df)=0$ diventa, tradotto nel linguaggio del Modulo 7, l'identità $\nabla\times(\nabla f)=0$ — **il rotore di un gradiente è sempre nullo**, un fatto che nel notebook 7.22 poteva sembrare una coincidenza di calcolo, e che qui si rivela essere un caso particolare di un unico principio algebrico. Applicando $d^2=0$ a una 1-forma $\omega$ associata a $F$: $d(d\omega)=0$ diventa $\nabla\cdot(\nabla\times F)=0$ — **la divergenza di un rotore è sempre nulla**, un'altra identità del Modulo 7 che ora appare come conseguenza dello stesso fatto algebrico $d^2=0$.

Questa unificazione è il primo grande guadagno delle forme differenziali: fenomeni che nel Modulo 7 sembravano tre argomenti scollegati (gradiente, rotore, divergenza) più due identità isolate, sono in realtà **un solo oggetto** ($d$) e **una sola proprietà** ($d^2=0$).

```python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)

# Verifica simbolica: rotore di un gradiente e' sempre zero (d^2 f = 0 sulle funzioni)
f = x**2*sp.sin(y) + x*z**3
grad_f = sp.Matrix([sp.diff(f, x), sp.diff(f, y), sp.diff(f, z)])

def rotore(F):
    P, Q, R = F
    return sp.Matrix([
        sp.diff(R, y) - sp.diff(Q, z),
        sp.diff(P, z) - sp.diff(R, x),
        sp.diff(Q, x) - sp.diff(P, y),
    ])

rot_grad_f = sp.simplify(rotore(grad_f))
print("rot(grad f) =", rot_grad_f.T, " (atteso: vettore nullo)")

# Verifica simbolica: divergenza di un rotore e' sempre zero (d^2 omega = 0 sulle 1-forme)
F = sp.Matrix([x*y*z, sp.sin(x*y), sp.exp(y)*z**2])
rot_F = rotore(F)

def divergenza(F):
    P, Q, R = F
    return sp.diff(P, x) + sp.diff(Q, y) + sp.diff(R, z)

div_rot_F = sp.simplify(divergenza(rot_F))
print("div(rot F) =", div_rot_F, " (atteso: 0)")
```

## Esercizio svolto e commentato

**Problema:** verifica, con un esempio concreto, che $d$ applicata alla 1-forma $\omega=P\,dx+Q\,dy+R\,dz$ dà davvero le componenti del rotore, con $P=y z, Q=xz, R=xy$ (il campo $F=\nabla(xyz)$, un gradiente — quindi ci aspettiamo $d\omega=0$, per l'identità appena vista).

**Ragionamento:**

**Passo 1 — riconosciamo che $F$ è un gradiente.** $F=(yz,xz,xy)=\nabla(xyz)$: infatti $\partial_x(xyz)=yz$, $\partial_y(xyz)=xz$, $\partial_z(xyz)=xy$ ✓. Quindi, per la teoria, ci aspettiamo $d\omega=0$ **prima ancora di calcolare**.

**Passo 2 — calcoliamo comunque i tre coefficienti di $d\omega$ per esercizio.** $d\omega=(R_y-Q_z)\,dy\wedge dz+(P_z-R_x)\,dz\wedge dx+(Q_x-P_y)\,dx\wedge dy$.

**Passo 3 — coefficiente di $dy\wedge dz$.** $R_y=\partial_y(xy)=x$, $Q_z=\partial_z(xz)=x$. $R_y-Q_z=x-x=0$.

**Passo 4 — coefficiente di $dz\wedge dx$.** $P_z=\partial_z(yz)=y$, $R_x=\partial_x(xy)=y$. $P_z-R_x=y-y=0$.

**Passo 5 — coefficiente di $dx\wedge dy$.** $Q_x=\partial_x(xz)=z$, $P_y=\partial_y(yz)=z$. $Q_x-P_y=z-z=0$.

**Conclusione:** $d\omega=0\cdot dy\wedge dz+0\cdot dz\wedge dx+0\cdot dx\wedge dy=0$, confermando **senza sorprese** quanto previsto al Passo 1 dall'identità $d(df)=0$: qui $\omega=df$ con $f=xyz$, quindi $d\omega=d(df)=0$ per il teorema generale, e il calcolo esplicito lo conferma componente per componente.

```python
import sympy as sp

x, y, z = sp.symbols('x y z', real=True)
P, Q, R = y*z, x*z, x*y   # F = grad(xyz)

domega_dydz = sp.diff(R, y) - sp.diff(Q, z)
domega_dzdx = sp.diff(P, z) - sp.diff(R, x)
domega_dxdy = sp.diff(Q, x) - sp.diff(P, y)

print("coefficiente di dy^dz:", sp.simplify(domega_dydz))
print("coefficiente di dz^dx:", sp.simplify(domega_dzdx))
print("coefficiente di dx^dy:", sp.simplify(domega_dxdy))
print("Tutti nulli, come previsto da d(df)=0 con f=xyz.")
```

Dunque $f(x,y,z)=x^2y+yz^2$ (a meno di una costante additiva), e si verifica direttamente $\nabla f=(2xy,x^2+z^2,2yz)=F$ ✓.

Per l'uguaglianza delle derivate miste (notebook 7.10), $f_{xy}=f_{yx}$, quindi il coefficiente è $0$: $d(df)=0$.

## Riepilogo finale

- Il **duale** $T_p^*$ dello spazio tangente $T_p$ contiene i covettori; il **differenziale** $df$ di una funzione è l'esempio più semplice, con le stesse componenti del gradiente ma un significato concettualmente distinto.
- Le **$k$-forme differenziali**, costruite con il prodotto esterno $\wedge$ antisimmetrico, generalizzano funzioni (0-forme), campi vettoriali visti come covettori (1-forme), e oggetti "di flusso" (2-forme, 3-forme, ...).
- La **derivata esterna** $d$ unifica gradiente ($d$ su 0-forme), rotore ($d$ su 1-forme in $\mathbb R^3$) e divergenza ($d$ su 2-forme in $\mathbb R^3$) in un'unica operazione, definita in qualunque dimensione.
- L'identità algebrica $d^2=0$ ricostruisce, come casi particolari, le identità $\nabla\times\nabla f=0$ e $\nabla\cdot(\nabla\times F)=0$ del Modulo 7, mostrando che non erano coincidenze isolate ma manifestazioni di un unico principio.
- Questo linguaggio unificato è quello che permetterà, nel notebook 16.11, di enunciare un **unico** teorema che contiene come casi particolari Green, Stokes e la divergenza (notebook 7.23-7.25).
