Manca un tassello: funzioni **vettoriali di una variabile**, $\gamma:I\to\mathbb R^n$, che descrivono il moto di un punto nello spazio — una traiettoria, una curva. Sono l'oggetto più semplice della geometria differenziale, perché richiedono solo la derivata a una variabile (Modulo 5), applicata componente per componente.

Questo notebook apre il Modulo 16, il cui obiettivo finale è generalizzare la geometria euclidea a spazi curvi. Le curve sono il primo passo naturale: prima di curvare uno spazio intero (una superficie, una varietà), impariamo a misurare quanto **una linea** si allontana dall'essere retta. Questa idea — la curvatura — sarà il filo conduttore di tutto il modulo, e qui la vediamo per la prima volta nel caso più semplice possibile.

## Teoria, passo dopo passo

### Curve parametrizzate

Una **curva parametrizzata** in $\mathbb R^n$ è una funzione $\gamma:I\to\mathbb R^n$, dove $I\subseteq\mathbb R$ è un intervallo, scritta in componenti $\gamma(t)=(\gamma_1(t),\ldots,\gamma_n(t))$. Pensiamo a $t$ come al tempo e a $\gamma(t)$ come alla posizione di un punto mobile. Esempi già impliciti in notebook precedenti: la retta $\gamma(t)=P+t\vec v$ (notebook 2.2, ora vista come curva), la circonferenza $\gamma(t)=(r\cos t,r\sin t)$ (notebook 2.4 e 2.13, coordinate polari), l'ellisse $\gamma(t)=(a\cos t,b\sin t)$ (notebook 2.6).

### Velocità e vettore tangente

Se ogni componente $\gamma_i(t)$ è derivabile (notebook 5.8), definiamo il **vettore velocità** $\gamma'(t)=(\gamma_1'(t),\ldots,\gamma_n'(t))$, derivando componente per componente. Geometricamente $\gamma'(t_0)$ è il vettore tangente alla curva nel punto $\gamma(t_0)$: la direzione istantanea del moto. Una curva è **regolare** se $\gamma'(t)\ne0$ per ogni $t\in I$ — condizione che esclude punti d'arresto o cuspidi, dove la nozione di "direzione tangente" non è ben definita.

La **retta tangente** alla curva in $\gamma(t_0)$ è $\ell(s)=\gamma(t_0)+s\,\gamma'(t_0)$, l'analogo per curve della retta tangente al grafico di una funzione (notebook 5.8).

### Lunghezza d'arco

Su un intervallo piccolo $[t,t+dt]$, la curva è ben approssimata dal segmento $\gamma(t)+\gamma'(t)\,dt$, di lunghezza $\|\gamma'(t)\|\,dt$ (norma euclidea, notebook 7.2). Sommando questi contributi infinitesimi (un integrale di Riemann, notebook 5.15), otteniamo la **lunghezza d'arco** da $t_0$ a $t_1$:
$$L(t_0,t_1)=\int_{t_0}^{t_1}\|\gamma'(t)\|\,dt.$$

**Teorema (invarianza per riparametrizzazione).** Se $\tau:[a,b]\to I$ è una funzione derivabile, crescente, con $\tau(a)=t_0,\tau(b)=t_1$, e $\tilde\gamma=\gamma\circ\tau$, allora $L(\tilde\gamma)=L(\gamma)$ sullo stesso tratto.

**Dimostrazione.** $\tilde\gamma'(s)=\gamma'(\tau(s))\,\tau'(s)$ (regola della catena, notebook 5.9, applicata componente per componente). Poiché $\tau'(s)>0$ (crescente): $\|\tilde\gamma'(s)\|=\|\gamma'(\tau(s))\|\,\tau'(s)$. Allora $\int_a^b\|\tilde\gamma'(s)\|\,ds=\int_a^b\|\gamma'(\tau(s))\|\,\tau'(s)\,ds=\int_{t_0}^{t_1}\|\gamma'(t)\|\,dt$, per il teorema di cambiamento di variabile nell'integrale (conseguenza del teorema fondamentale del calcolo, notebook 5.16, applicato a $t=\tau(s)$). $\blacksquare$

Questo teorema dice una cosa importante: la lunghezza è una proprietà **della traccia geometrica**, non del modo particolare in cui la percorriamo. Per questo, per una curva regolare esiste sempre una **parametrizzazione per lunghezza d'arco**: scegliendo $s(t)=\int_{t_0}^t\|\gamma'(u)\|\,du$ (funzione crescente, per regolarità, quindi invertibile, notebook 3.6) e riparametrizzando con $t=s^{-1}$, si ottiene $\tilde\gamma(s)$ con $\|\tilde\gamma'(s)\|=1$ per ogni $s$: percorriamo la curva a velocità unitaria.

### Curvatura

Supponiamo $\gamma(s)$ parametrizzata per lunghezza d'arco, cioè $\|\gamma'(s)\|=1$. Il vettore tangente unitario è $T(s)=\gamma'(s)$. La **curvatura** misura quanto rapidamente $T$ cambia direzione:
$$\kappa(s)=\|T'(s)\|=\|\gamma''(s)\|.$$

Una retta ha $T$ costante, quindi $\kappa\equiv0$: coerente con l'idea che la curvatura misura lo scostamento dall'essere retta. Una circonferenza di raggio $r$, parametrizzata per lunghezza d'arco come $\gamma(s)=r(\cos(s/r),\sin(s/r))$, ha $\gamma''(s)=-\frac1r(\cos(s/r),\sin(s/r))$, quindi $\kappa=1/r$: **più piccolo è il raggio, più la circonferenza è "curva"** — coerente con l'intuizione.

**Fatto (curvatura da parametrizzazione qualunque).** Per una curva piana regolare $\gamma(t)=(x(t),y(t))$, non necessariamente per lunghezza d'arco, vale
$$\kappa(t)=\frac{x'(t)y''(t)-x''(t)y'(t)}{\big(x'(t)^2+y'(t)^2\big)^{3/2}},$$
formula che useremo nella verifica computazionale (si ottiene applicando la regola della catena alla definizione precedente, un calcolo che omettiamo qui).

### Il caso $\mathbb R^3$: torsione e formule di Frenet-Serret

In $\mathbb R^3$, con $\gamma$ parametrizzata per lunghezza d'arco e $\kappa(s)\ne0$, definiamo, oltre al versore tangente $T=\gamma'$:
- il **versore normale** $N=T'/\|T'\|=T'/\kappa$ (ben definito perché $\kappa\ne0$);
- il **versore binormale** $B=T\times N$ (prodotto vettoriale, notebook 7.19), che completa $\{T,N,B\}$ a una base ortonormale (notebook 6.22) di $\mathbb R^3$ in ogni punto — il **triedro di Frenet**.

La **torsione** $\tau(s)$ misura quanto la curva esce dal piano osculatore (il piano generato da $T,N$): si definisce tramite $B'=-\tau N$. Le tre derivate $T',N',B'$ si esprimono nella base $\{T,N,B\}$ stessa attraverso le **equazioni di Frenet-Serret**:
$$T'=\kappa N,\qquad N'=-\kappa T+\tau B,\qquad B'=-\tau N.$$
(La prima è la definizione di $N$; la terza è la definizione di $\tau$; la seconda si ricava derivando $N=B\times T$ e usando le altre due, un calcolo che il lettore può svolgere come esercizio.)

**Teorema fondamentale delle curve (enunciato).** Date due funzioni continue $\kappa(s)>0$ e $\tau(s)$ su un intervallo, esiste — a meno di rotazioni e traslazioni rigide dello spazio (isometrie di $\mathbb R^3$) — un'unica curva $\gamma(s)$ parametrizzata per lunghezza d'arco con curvatura $\kappa(s)$ e torsione $\tau(s)$ assegnate. Questo teorema (dimostrabile con il teorema di esistenza e unicità per sistemi di equazioni differenziali lineari — che generalizza a sistemi il teorema di esistenza e unicità per il problema di Cauchy scalare, notebook 8.1 — applicato al sistema di Frenet-Serret) dice che **curvatura e torsione determinano completamente la forma di una curva nello spazio**, a meno di dove e come la mettiamo: sono le sue "vere" caratteristiche geometriche intrinseche. Questa idea — separare le proprietà intrinseche di un oggetto geometrico dal modo in cui è immerso nello spazio ambiente — riapparirà in forma molto più profonda nel notebook 16.9, con la distinzione tra curvatura intrinseca ed estrinseca delle superfici.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Elica cilindrica: gamma(t) = (cos t, sin t, t)
t = np.linspace(0, 4*np.pi, 400)
x, y, z = np.cos(t), np.sin(t), t

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="elica $\\gamma(t)=(\\cos t,\\sin t, t)$")
ax.legend()
ax.set_title("Una curva regolare in $\\mathbb{R}^3$")
plt.tight_layout()
plt.show()

# Verifica della formula della curvatura per una curva piana: l'ellisse
# gamma(t) = (a cos t, b sin t)
a, b = 2.0, 1.0
tt = np.linspace(0, 2*np.pi, 200)
xp, yp = -a*np.sin(tt), b*np.cos(tt)
xpp, ypp = -a*np.cos(tt), -b*np.sin(tt)
kappa = (xp*ypp - xpp*yp) / (xp**2 + yp**2)**1.5

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(tt, kappa)
ax.set_xlabel("t")
ax.set_ylabel("curvatura $\\kappa(t)$")
ax.set_title(f"Curvatura dell'ellisse a={a}, b={b}: massima ai vertici sull'asse corto")
plt.tight_layout()
plt.show()
print("kappa massima:", kappa.max(), " kappa minima:", kappa.min())
print("Per confronto, la circonferenza di raggio b=1 avrebbe kappa=1 costante:", 1/b)
```

## Esercizio svolto e commentato

**Problema:** calcola la curvatura della circonferenza $\gamma(t)=(r\cos t,r\sin t)$ usando la formula generale (non la parametrizzazione per lunghezza d'arco), e verifica che coincide con $1/r$ trovato nella teoria.

**Ragionamento:**

**Passo 1 — perché non usiamo direttamente la parametrizzazione per lunghezza d'arco?** Perché vogliamo verificare la formula generale $\kappa(t)=\dfrac{x'y''-x''y'}{(x'^2+y'^2)^{3/2}}$ su un caso di cui già conosciamo la risposta: un buon modo per fidarsi di una formula nuova prima di usarla su casi ignoti.

**Passo 2 — calcoliamo le derivate.** $x(t)=r\cos t\Rightarrow x'=-r\sin t,\ x''=-r\cos t$. $y(t)=r\sin t\Rightarrow y'=r\cos t,\ y''=-r\sin t$.

**Passo 3 — sostituiamo nel numeratore.** $x'y''-x''y'=(-r\sin t)(-r\sin t)-(-r\cos t)(r\cos t)=r^2\sin^2t+r^2\cos^2t=r^2$ (identità $\sin^2+\cos^2=1$, Modulo 2).

**Passo 4 — sostituiamo nel denominatore.** $x'^2+y'^2=r^2\sin^2t+r^2\cos^2t=r^2$, quindi $(x'^2+y'^2)^{3/2}=r^3$ (assumendo $r>0$).

**Passo 5 — concludiamo.** $\kappa(t)=\dfrac{r^2}{r^3}=\dfrac1r$, **costante**, indipendente da $t$: coincide esattamente con il valore trovato nella teoria usando la parametrizzazione per lunghezza d'arco. Questo conferma che la curvatura è una proprietà **geometrica** della curva (non dipende da come la percorriamo), coerentemente col teorema di invarianza per riparametrizzazione visto per la lunghezza — un fatto analogo vale per la curvatura, anche se qui lo abbiamo solo verificato su un esempio, non dimostrato in generale.

```python
import sympy as sp

t, r = sp.symbols('t r', positive=True, real=True)
x = r*sp.cos(t)
y = r*sp.sin(t)

xp, yp = sp.diff(x, t), sp.diff(y, t)
xpp, ypp = sp.diff(xp, t), sp.diff(yp, t)

kappa = sp.simplify((xp*ypp - xpp*yp) / (xp**2 + yp**2)**sp.Rational(3, 2))
print("Curvatura della circonferenza di raggio r:", kappa)
assert kappa == 1/r
print("Confermato: kappa = 1/r, indipendente da t.")
```

Per $c\ne0$: $s(t)=\int_0^t\sqrt{1+c^2}\,du=t\sqrt{1+c^2}$, quindi la riparametrizzazione per lunghezza d'arco è $t=s/\sqrt{1+c^2}$, e $\tilde\gamma(s)=\big(\cos(s/\sqrt{1+c^2}),\sin(s/\sqrt{1+c^2}),cs/\sqrt{1+c^2}\big)$.

Sostituendo $B'=-\tau N$ e $T'=\kappa N$: $N'=(-\tau N)\times T+B\times(\kappa N)=-\tau(N\times T)+\kappa(B\times N)$. Usando le regole del prodotto vettoriale sulla base ortonormale destrorsa $\{T,N,B\}$ (con $T\times N=B$, quindi $N\times T=-B$, e $B\times N=-T$ dato che $N\times B=T$): $N'=-\tau(-B)+\kappa(-T)=-\kappa T+\tau B$. Esattamente la seconda equazione di Frenet-Serret.

## Riepilogo finale

- Una curva parametrizzata $\gamma:I\to\mathbb R^n$ ha vettore velocità $\gamma'(t)$, ottenuto derivando componente per componente (Modulo 5); è regolare se $\gamma'(t)\ne0$ ovunque.
- La lunghezza d'arco $L=\int\|\gamma'(t)\|\,dt$ è invariante per riparametrizzazione: è una proprietà della traccia geometrica, non della velocità con cui la percorriamo.
- La curvatura $\kappa=\|T'(s)\|$ (con $T$ tangente unitario, $s$ parametro-lunghezza-d'arco) misura quanto la curva si scosta dall'essere retta: $\kappa\equiv0$ caratterizza esattamente le rette.
- Nello spazio, il triedro di Frenet $\{T,N,B\}$ e le equazioni di Frenet-Serret codificano completamente la forma della curva tramite curvatura e torsione: il teorema fondamentale delle curve mostra che queste due funzioni la determinano a meno di isometrie rigide.
- L'idea di separare le proprietà "vere" (geometriche, intrinseche) di un oggetto da come è disegnato o immerso nello spazio è il primo assaggio di un tema che dominerà l'intero Modulo 16.
