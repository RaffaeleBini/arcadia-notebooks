## Perché ci interessa

Il notebook 16.6 ha introdotto la distanza riemanniana $d_g(p,q)$ come estremo inferiore delle lunghezze di curve che uniscono $p$ a $q$. Una domanda naturale: **quali curve realizzano effettivamente questo minimo?** Su $\mathbb R^n$ con la metrica euclidea, la risposta è ovvia: il segmento di retta. Ma su una sfera, o su una varietà curva qualunque, cosa gioca il ruolo delle "rette"?

Questo notebook risponde tramite un **principio variazionale**: invece di minimizzare una funzione di un numero finito di variabili (come nel notebook 7.12, punti di massimo/minimo), minimizziamo un **funzionale** — l'integrale di lunghezza — su uno spazio di curve, infinito-dimensionale. Le curve che minimizzano (localmente) questo funzionale si chiamano **geodetiche**: sono le "linee più dritte possibili" che la geometria della varietà permette.

## Teoria, passo dopo passo

### Dall'ottimizzazione di funzioni all'ottimizzazione di curve

Nel notebook 7.12, per trovare i punti critici di una funzione $f(x_1,\ldots,x_n)$, imponevamo $\nabla f=0$: annullare la derivata rispetto a **ogni singola variazione infinitesima** delle coordinate. L'idea si estende a un funzionale $J[\gamma]$ che dipende da un'intera curva $\gamma$: cerchiamo curve $\gamma$ tali che, per ogni "variazione infinitesima" $\gamma_s=\gamma+s\,\eta$ (con $\eta$ una perturbazione arbitraria che si annulla agli estremi, per tenere fissi gli estremi della curva), si abbia $\dfrac{d}{ds}\Big|_{s=0}J[\gamma_s]=0$.

### L'equazione di Eulero-Lagrange

Per semplicità lavoriamo in coordinate locali su una varietà riemanniana $(M,g)$ (notebook 16.6), con $\gamma(t)=(x^1(t),\ldots,x^n(t))$ e il funzionale energia (equivalente, per i punti critici, al funzionale lunghezza, come vedremo)
$$J[\gamma]=\int_a^bg_{ij}(\gamma(t))\,\dot x^i(t)\dot x^j(t)\,dt$$
(qui e nel seguito, sommiamo automaticamente sugli indici ripetuti $i,j$, una convenzione — la **convenzione di Einstein** — che accorcia le formule; $\dot x^i$ indica $dx^i/dt$). Scrivendo $\mathcal L(x,\dot x)=g_{ij}(x)\dot x^i\dot x^j$ (una funzione, detta Lagrangiana, delle $2n$ variabili posizione-velocità), il calcolo delle variazioni (una generalizzazione dei moltiplicatori di Lagrange del notebook 7.13, applicata non più a un vincolo su un numero finito di variabili ma a un funzionale integrale) mostra che i punti critici di $J$ soddisfano, per ogni indice $k$, l'**equazione di Eulero-Lagrange**:
$$\frac{d}{dt}\Big(\frac{\partial\mathcal L}{\partial\dot x^k}\Big)-\frac{\partial\mathcal L}{\partial x^k}=0.$$

**Perché questa equazione.** L'idea, analoga al notebook 7.12, è che in un punto critico la derivata rispetto a ogni variazione infinitesima deve annullarsi. Per una perturbazione $\eta$ nulla agli estremi, $\dfrac{d}{ds}\Big|_0J[\gamma+s\eta]=\displaystyle\int_a^b\Big(\frac{\partial\mathcal L}{\partial x^k}\eta^k+\frac{\partial\mathcal L}{\partial\dot x^k}\dot\eta^k\Big)dt$ (regola della catena, notebook 7.20, applicata a $\mathcal L(\gamma(t)+s\eta(t),\dot\gamma(t)+s\dot\eta(t))$). Integrando per parti (notebook 5.17) il secondo termine, e usando $\eta(a)=\eta(b)=0$ per eliminare il termine di bordo: $=\displaystyle\int_a^b\Big(\frac{\partial\mathcal L}{\partial x^k}-\frac{d}{dt}\frac{\partial\mathcal L}{\partial\dot x^k}\Big)\eta^k\,dt$. Perché questo si annulli per **ogni** scelta di $\eta$ (arbitraria, nulla solo agli estremi), il fattore tra parentesi deve essere identicamente nullo — esattamente l'equazione di Eulero-Lagrange.

### L'equazione della geodetica

Applicando l'equazione di Eulero-Lagrange a $\mathcal L=g_{ij}\dot x^i\dot x^j$: $\dfrac{\partial\mathcal L}{\partial\dot x^k}=2g_{kj}\dot x^j$ (usando la simmetria di $g_{ij}$), e $\dfrac{\partial\mathcal L}{\partial x^k}=\dfrac{\partial g_{ij}}{\partial x^k}\dot x^i\dot x^j$. Dopo alcuni passaggi algebrici (che coinvolgono derivare $g_{kj}(x(t))$ rispetto a $t$ con la regola della catena, e poi risolvere per $\ddot x^k$ moltiplicando per la matrice inversa $g^{k\ell}$, che esiste perché $g$ è definita positiva quindi invertibile, notebook 6.16), si ottiene la forma standard dell'**equazione della geodetica**:
$$\ddot x^k+\Gamma^k_{ij}\dot x^i\dot x^j=0,$$
dove i coefficienti $\Gamma^k_{ij}$ (i **simboli di Christoffel**, che definiremo esplicitamente in termini di $g$ nel notebook 16.8) dipendono solo dalla metrica e dalle sue derivate prime. Questa è un **sistema di equazioni differenziali del secondo ordine** (notebook 8.5): dato un punto iniziale $x(0)=p$ e una velocità iniziale $\dot x(0)=v$, il teorema di esistenza e unicità per sistemi di EDO (che generalizza a sistemi non lineari il teorema di esistenza e unicità per il problema di Cauchy scalare, notebook 8.1) garantisce che esiste un'unica geodetica (almeno per un breve intervallo di tempo) che parte da $p$ con velocità $v$ — l'analogo, per curve su una varietà, del fatto che per un punto e una direzione passa un'unica retta.

### Perché il funzionale energia e non direttamente la lunghezza

Si potrebbe pensare di applicare Eulero-Lagrange direttamente al funzionale lunghezza $L[\gamma]=\int\sqrt{g_{ij}\dot x^i\dot x^j}\,dt$ del notebook 16.6. Si ottengono le stesse curve come punti critici, ma l'equazione risultante è più complicata (contiene $\|\dot\gamma\|$ al denominatore) perché $L$ è invariante per riparametrizzazione (notebook 16.1: la lunghezza non dipende da come percorriamo la curva), quindi non seleziona un'unica parametrizzazione. Il funzionale energia $J$, non essendo invariante per riparametrizzazione, seleziona automaticamente le geodetiche **parametrizzate a velocità costante** ($g_{ij}\dot x^i\dot x^j=$ costante lungo la soluzione, un fatto verificabile derivando questa quantità rispetto a $t$ e usando l'equazione della geodetica) — la parametrizzazione più naturale per i calcoli successivi.

### Esempi

- **$\mathbb R^n$ con la metrica euclidea:** $g_{ij}=\delta_{ij}$ costante, quindi $\Gamma^k_{ij}=0$ ovunque (nessuna derivata di $g$), e l'equazione della geodetica si riduce a $\ddot x^k=0$: le soluzioni sono $x(t)=p+tv$, **rette**, coerentemente con l'intuizione.
- **Sfera $S^2$ con la metrica indotta:** le geodetiche sono i **cerchi massimi** (circonferenze ottenute intersecando la sfera con un piano per il centro) — un fatto che si può verificare direttamente sostituendo la parametrizzazione di un cerchio massimo nell'equazione della geodetica per la metrica sferica del notebook 16.6, o più elegantemente osservando che un cerchio massimo è l'intersezione della sfera con un piano di simmetria, e la geodetica, per unicità, deve rispettare quella simmetria.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from mpl_toolkits.mplot3d import Axes3D

# Risolviamo numericamente l'equazione della geodetica sulla sfera S^2
# (parametrizzata da theta, phi, con metrica g_thetatheta=1, g_phiphi=sin^2(theta))
# I simboli di Christoffel non nulli per questa metrica sono:
#   Gamma^theta_phiphi = -sin(theta)cos(theta)
#   Gamma^phi_thetaphi = Gamma^phi_phitheta = cos(theta)/sin(theta)

def geodetica_sfera(t, y):
    theta, phi, dtheta, dphi = y
    ddtheta = np.sin(theta) * np.cos(theta) * dphi**2
    ddphi = -2 * (np.cos(theta) / np.sin(theta)) * dtheta * dphi
    return [dtheta, dphi, ddtheta, ddphi]

# Condizioni iniziali: partiamo dall'equatore, muovendoci con una componente in theta
y0 = [np.pi/2, 0.0, 0.5, 0.7]
sol = solve_ivp(geodetica_sfera, [0, 6], y0, dense_output=True, max_step=0.01)

t_vals = np.linspace(0, 6, 400)
theta_vals, phi_vals = sol.sol(t_vals)[0], sol.sol(t_vals)[1]

X = np.sin(theta_vals) * np.cos(phi_vals)
Y = np.sin(theta_vals) * np.sin(phi_vals)
Z = np.cos(theta_vals)

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')

# la sfera come sfondo
u, v = np.meshgrid(np.linspace(0, np.pi, 30), np.linspace(0, 2*np.pi, 30))
ax.plot_surface(np.sin(u)*np.cos(v), np.sin(u)*np.sin(v), np.cos(u), alpha=0.15, color='gray')
ax.plot(X, Y, Z, color='red', linewidth=2, label='geodetica numerica')
ax.legend()
ax.set_title("Geodetica sulla sfera: un cerchio massimo")
plt.tight_layout()
plt.show()

# Verifica: un cerchio massimo giace su un piano per l'origine.
# Controlliamo che i punti della geodetica soddisfino (quasi) un'equazione lineare a*X+b*Y+c*Z=0
# trovando il piano tramite i primi due punti e l'origine (prodotto vettoriale)
p1 = np.array([X[0], Y[0], Z[0]])
p2 = np.array([X[50], Y[50], Z[50]])
normale = np.cross(p1, p2)
normale = normale / np.linalg.norm(normale)
residui = normale[0]*X + normale[1]*Y + normale[2]*Z
print("Errore massimo dal piano del cerchio massimo:", np.max(np.abs(residui)))
```

## Esercizio svolto e commentato

**Problema:** verifica esplicitamente che l'equatore $\theta(t)=\pi/2$ costante, $\phi(t)=t$ è una geodetica della sfera, usando i simboli di Christoffel dati nella cella di codice ($\Gamma^\theta_{\phi\phi}=-\sin\theta\cos\theta$, $\Gamma^\phi_{\theta\phi}=\Gamma^\phi_{\phi\theta}=\cot\theta$, tutti gli altri nulli).

**Ragionamento:**

**Passo 1 — scriviamo le due componenti dell'equazione della geodetica.** $\ddot\theta+\Gamma^\theta_{\phi\phi}\dot\phi^2=0$ (l'unico termine con $\Gamma^\theta$ non nullo) e $\ddot\phi+2\Gamma^\phi_{\theta\phi}\dot\theta\dot\phi=0$ (il fattore $2$ viene dal fatto che $\Gamma^\phi_{\theta\phi}=\Gamma^\phi_{\phi\theta}$ compare due volte nella somma su $i,j$ della formula generale).

**Passo 2 — calcoliamo le derivate della curva candidata.** $\theta(t)=\pi/2\Rightarrow\dot\theta=0,\ddot\theta=0$. $\phi(t)=t\Rightarrow\dot\phi=1,\ddot\phi=0$.

**Passo 3 — verifichiamo la prima equazione.** $\ddot\theta+\Gamma^\theta_{\phi\phi}\dot\phi^2=0+(-\sin(\pi/2)\cos(\pi/2))\cdot1^2=-1\cdot0\cdot1=0$ ✓ (usando $\cos(\pi/2)=0$).

**Passo 4 — verifichiamo la seconda equazione.** $\ddot\phi+2\Gamma^\phi_{\theta\phi}\dot\theta\dot\phi=0+2\cot(\pi/2)\cdot0\cdot1=0$ (il fattore $\dot\theta=0$ annulla tutto, indipendentemente dal valore di $\cot(\pi/2)=0$) ✓.

**Conclusione:** entrambe le componenti dell'equazione della geodetica sono soddisfatte identicamente: l'equatore è davvero una geodetica, coerentemente con il fatto che è un cerchio massimo (intersezione della sfera con il piano $z=0$, che passa per il centro). Questo conferma un caso particolare, verificabile a mano, del risultato generale osservato numericamente nella cella di codice per una geodetica generica.

```python
import sympy as sp

t = sp.symbols('t', real=True)
theta = sp.pi/2  # costante
phi = t

dtheta = 0  # theta e' costante = pi/2
ddtheta = 0
dphi = sp.diff(phi, t)
ddphi = sp.diff(phi, t, 2)

Gamma_theta_phiphi = -sp.sin(theta)*sp.cos(theta)
Gamma_phi_thetaphi = sp.cos(theta)/sp.sin(theta)

eq1 = ddtheta + Gamma_theta_phiphi * dphi**2
eq2 = ddphi + 2*Gamma_phi_thetaphi*dtheta*dphi

print("Equazione 1 (deve essere 0):", sp.simplify(eq1))
print("Equazione 2 (deve essere 0):", sp.simplify(eq2))
```

**4.** $\dot E=\partial_kg_{ij}\dot x^k\dot x^i\dot x^j+2g_{ij}\ddot x^i\dot x^j$ (regola della catena su $g_{ij}(\gamma(t))$, più regola del prodotto sul prodotto $\dot x^i\dot x^j$, usando la simmetria $g_{ij}=g_{ji}$ per combinare i due termini in $2g_{ij}\ddot x^i\dot x^j$). Sostituendo $\ddot x^i=-\Gamma^i_{k\ell}\dot x^k\dot x^\ell$ dall'equazione della geodetica: $\dot E=\partial_kg_{ij}\dot x^k\dot x^i\dot x^j-2g_{ij}\Gamma^i_{k\ell}\dot x^k\dot x^\ell\dot x^j$. ## Riepilogo finale

- Le **geodetiche** sono i punti critici del funzionale lunghezza (o, equivalentemente, del funzionale energia) tra tutte le curve che uniscono due punti fissati: la generalizzazione riemanniana delle rette.
- L'equazione di Eulero-Lagrange, ottenuta annullando la derivata del funzionale rispetto a ogni variazione infinitesima della curva, porta all'**equazione della geodetica** $\ddot x^k+\Gamma^k_{ij}\dot x^i\dot x^j=0$, un sistema di EDO del secondo ordine.
- In $\mathbb R^n$ euclideo, dove $\Gamma^k_{ij}=0$, le geodetiche sono esattamente le rette; sulla sfera sono i cerchi massimi.
- Il teorema di esistenza e unicità per sistemi di EDO (generalizzazione a sistemi non lineari del teorema di Cauchy scalare, notebook 8.1) garantisce che, per ogni punto e ogni direzione iniziale, esiste un'unica geodetica: l'analogo riemanniano del fatto euclideo che per un punto e una direzione passa un'unica retta.
- L'energia cinetica $g_{ij}\dot x^i\dot x^j$ è conservata lungo le geodetiche: si muovono a "velocità costante".
