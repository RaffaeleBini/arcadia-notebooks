## Perché ci interessa

Il notebook 16.2 ha mostrato che, su una superficie immersa in $\mathbb R^3$, il prodotto scalare di $\mathbb R^3$ ristretto al piano tangente (la prima forma fondamentale) permette di misurare lunghezze e angoli **usando solo dati interni alla superficie**. Ma su una varietà astratta (notebook 16.5), non c'è nessun $\mathbb R^3$ ambiente da cui ereditare un prodotto scalare: bisogna **sceglierne uno** su ogni spazio tangente, in modo coerente da punto a punto.

Questa scelta si chiama **metrica riemanniana**, ed è l'oggetto centrale di tutta la geometria differenziale: una volta fissata una metrica, ha senso parlare di lunghezze, angoli, aree, e — nei prossimi notebook — di "linee rette" (geodetiche) e di curvatura, tutto senza bisogno di alcuno spazio ambiente.

## Teoria, passo dopo passo

### Definizione di metrica riemanniana

Una **metrica riemanniana** $g$ su una varietà differenziabile $M$ (notebook 16.5) è una scelta, per ogni $p\in M$, di un prodotto scalare $g_p:T_pM\times T_pM\to\mathbb R$ (notebook 6.20: bilineare, simmetrico, definito positivo) sullo spazio tangente in $p$, che varia in modo liscio da punto a punto (cioè, in coordinate locali, le componenti di $g_p$ sono funzioni lisce di $p$). La coppia $(M,g)$ si chiama **varietà riemanniana**.

In una carta $(U,\varphi)$ con coordinate $(x_1,\ldots,x_n)$, usando la base $\Big\{\dfrac{\partial}{\partial x_i}\Big|_p\Big\}$ di $T_pM$ (notebook 16.3), la metrica è determinata dalle sue componenti
$$g_{ij}(p)=g_p\Big(\frac{\partial}{\partial x_i}\Big|_p,\frac{\partial}{\partial x_j}\Big|_p\Big),$$
una matrice simmetrica (notebook 6.23) e definita positiva (notebook 6.17-6.19: autovalori tutti positivi) in ogni punto. Per due vettori tangenti $v=\sum v^i\partial_i,w=\sum w^j\partial_j$: $g_p(v,w)=\sum_{i,j}g_{ij}(p)\,v^iw^j$ — una forma bilineare simmetrica esattamente come nel notebook 6.20, applicata punto per punto.

### La metrica come generalizzazione della prima forma fondamentale

Per una superficie $\Sigma=\sigma(U)\subset\mathbb R^3$ (notebook 16.2), la metrica indotta è esattamente $g_{11}=E,\ g_{12}=g_{21}=F,\ g_{22}=G$: la prima forma fondamentale **è** una metrica riemanniana, nel caso particolare in cui esiste uno spazio ambiente da cui ereditarla. La differenza concettuale è che ora, su una varietà astratta, **nessuno spazio ambiente fornisce automaticamente una metrica**: dobbiamo sceglierla, e scelte diverse sulla stessa varietà topologica danno geometrie diverse (lunghezze, angoli, e — vedremo nel notebook 16.9 — curvature diverse).

### Lunghezza di curve e distanza

Data una curva $\gamma:[a,b]\to M$ (una curva su una varietà, definita esattamente come nel notebook 16.1 ma con valori in $M$ invece che in $\mathbb R^n$, usando le carte per dare senso alla derivabilità), la sua **lunghezza** rispetto a $g$ è
$$L_g(\gamma)=\int_a^b\sqrt{g_{\gamma(t)}(\gamma'(t),\gamma'(t))}\,dt,$$
esattamente la formula del notebook 16.2 (lì con $E,F,G$), ora scritta con $g_{ij}$ generici. La **distanza riemanniana** tra due punti $p,q\in M$ si definisce come $d_g(p,q)=\inf_\gamma L_g(\gamma)$, l'estremo inferiore (notebook 5.2, completezza dei reali, per l'esistenza dell'estremo inferiore) sulle curve $\gamma$ che uniscono $p$ a $q$. Si dimostra che $d_g$ soddisfa gli assiomi di uno spazio metrico (notebook 12.7): $d_g(p,q)\ge0$ con uguaglianza solo se $p=q$, simmetria, e la disuguaglianza triangolare (quest'ultima segue dal fatto che concatenare due curve dà una curva la cui lunghezza è la somma delle lunghezze, e l'estremo inferiore su un insieme più ampio di curve non può che essere minore o uguale). $(M,d_g)$ è quindi, in particolare, uno **spazio metrico** nel senso del Modulo 12, con tutta la teoria di quel modulo (completezza, notebook 12.8; compattezza, notebook 12.6) applicabile.

### Esempi

- **Metrica euclidea su $\mathbb R^n$:** $g_{ij}=\delta_{ij}$ (identità), da cui $L_g(\gamma)=\int\|\gamma'(t)\|\,dt$, la lunghezza ordinaria del notebook 16.1.
- **Metrica sferica su $S^2$**, indotta da $\mathbb R^3$: nelle coordinate $(\theta,\phi)$ del notebook 16.2, $g_{\theta\theta}=r^2,g_{\theta\phi}=0,g_{\phi\phi}=r^2\sin^2\theta$ (esattamente $E,F,G$ calcolati lì per $r=1$).
- **Metrica iperbolica sul disco di Poincaré**, $D=\{(x,y):x^2+y^2<1\}$: $g_{ij}=\dfrac{4}{(1-x^2-y^2)^2}\delta_{ij}$ — una metrica **non indotta da nessun'immersione in $\mathbb R^3$** con la geometria euclidea usuale, ma perfettamente legittima come metrica riemanniana astratta. Man mano che $(x,y)\to$ bordo del disco, il fattore $\frac{4}{(1-x^2-y^2)^2}\to\infty$: le lunghezze esplodono vicino al bordo, che risulta "infinitamente lontano" nella distanza $d_g$, pur essendo a distanza euclidea finita — un esempio in cui l'intuizione euclidea sulle distanze fallisce completamente, ma la teoria riemanniana resta perfettamente coerente.

```python
import numpy as np
import matplotlib.pyplot as plt

# Confrontiamo lunghezze di curve nel disco di Poincare' rispetto alla metrica
# iperbolica g_ij = 4/(1-x^2-y^2)^2 * delta_ij, contro la lunghezza euclidea.

def fattore_conforme(x, y):
    return 4.0 / (1 - x**2 - y**2)**2

def lunghezza_curva(x_t, y_t, t, metrica='iperbolica'):
    dx = np.gradient(x_t, t)
    dy = np.gradient(y_t, t)
    speed_euclidea = np.sqrt(dx**2 + dy**2)
    if metrica == 'euclidea':
        return np.trapz(speed_euclidea, t)
    fattore = np.sqrt(fattore_conforme(x_t, y_t))
    return np.trapz(fattore * speed_euclidea, t)

# Segmento radiale dal centro verso il bordo (fino a raggio 0.99)
t = np.linspace(0, 1, 2000)
x_t = 0.99 * t
y_t = np.zeros_like(t)

L_eucl = lunghezza_curva(x_t, y_t, t, 'euclidea')
L_iperb = lunghezza_curva(x_t, y_t, t, 'iperbolica')

print(f"Lunghezza euclidea del segmento verso il bordo: {L_eucl:.4f} (finita, ovviamente)")
print(f"Lunghezza iperbolica dello stesso segmento: {L_iperb:.4f} (molto piu' grande)")

# Facciamo vedere che, avvicinandosi sempre di piu' al bordo, la lunghezza iperbolica diverge
raggi_massimi = [0.5, 0.9, 0.99, 0.999, 0.9999]
for r_max in raggi_massimi:
    t = np.linspace(0, 1, 5000)
    x_t = r_max * t
    y_t = np.zeros_like(t)
    L = lunghezza_curva(x_t, y_t, t, 'iperbolica')
    print(f"raggio massimo {r_max}: lunghezza iperbolica = {L:.3f}")
```

## Esercizio svolto e commentato

**Problema:** calcola la lunghezza iperbolica esatta (con SymPy) del segmento radiale da $0$ a $r_0<1$ nel disco di Poincaré, e verifica che tende a $+\infty$ per $r_0\to1^-$.

**Ragionamento:**

**Passo 1 — parametrizziamo il segmento.** $\gamma(t)=(t,0)$ per $t\in[0,r_0]$, quindi $\gamma'(t)=(1,0)$, con $\|\gamma'(t)\|_{\text{eucl}}=1$.

**Passo 2 — scriviamo l'integrale di lunghezza iperbolica.** $g_{ij}=\dfrac4{(1-x^2-y^2)^2}\delta_{ij}$, quindi lungo $\gamma$ (dove $y=0$): $g_{\gamma(t)}(\gamma'(t),\gamma'(t))=\dfrac4{(1-t^2)^2}\cdot1$. La lunghezza è
$$L=\int_0^{r_0}\sqrt{\frac4{(1-t^2)^2}}\,dt=\int_0^{r_0}\frac2{1-t^2}\,dt.$$

**Passo 3 — calcoliamo l'integrale (tecnica di integrazione, notebook 5.17: fratti semplici).** $\dfrac2{1-t^2}=\dfrac1{1-t}+\dfrac1{1+t}$ (si verifica sommando le frazioni). Quindi $L=\Big[-\ln(1-t)+\ln(1+t)\Big]_0^{r_0}=\ln\dfrac{1+r_0}{1-r_0}$.

**Passo 4 — verifichiamo la divergenza per $r_0\to1^-$.** Quando $r_0\to1^-$, il numeratore $1+r_0\to2$ (finito) ma il denominatore $1-r_0\to0^+$: il rapporto $\to+\infty$, quindi $L\to+\infty$ (per la continuità del logaritmo, notebook 3.14, composta con un limite che tende a $+\infty$, notebook 5.4). Conclusione: **il bordo del disco è a distanza iperbolica infinita da ogni punto interno**, confermando numericamente quanto osservato nella cella di verifica computazionale (dove $r_{\max}=0.9999$ già dava una lunghezza molto grande, coerente con $\ln\frac{1.9999}{0.0001}\approx\ln(19999)\approx9.9$).

```python
import sympy as sp

t, r0 = sp.symbols('t r0', real=True, positive=True)
integrando = 2 / (1 - t**2)
L = sp.integrate(integrando, (t, 0, r0))
L_semplificata = sp.simplify(L)
print("Lunghezza iperbolica del segmento [0, r0]:", L_semplificata)

# confronto con la formula trovata a mano: ln((1+r0)/(1-r0))
formula_attesa = sp.log((1+r0)/(1-r0))
print("Coincide con ln((1+r0)/(1-r0))?", sp.simplify(L_semplificata - formula_attesa) == 0)

# limite per r0 -> 1
limite = sp.limit(formula_attesa, r0, 1, dir='-')
print("Limite per r0 -> 1^-:", limite)
```

## Riepilogo finale

- Una **metrica riemanniana** $g$ è una scelta liscia di prodotto scalare $g_p$ su ogni spazio tangente $T_pM$, con componenti $g_{ij}$ in coordinate locali: generalizza la prima forma fondamentale del notebook 16.2 a varietà astratte, senza bisogno di uno spazio ambiente.
- Con $g$ si definisce la lunghezza di curve $L_g(\gamma)$ e, per estremo inferiore, la **distanza riemanniana** $d_g$, che rende $(M,d_g)$ uno spazio metrico nel senso del Modulo 12.
- Esempi diversi (euclideo, sferico, iperbolico) mostrano come scelte diverse di $g$ sulla stessa varietà topologica producano geometrie radicalmente diverse: nel disco di Poincaré il bordo è a distanza infinita, un fenomeno impossibile nella metrica euclidea.
- Con una metrica in mano, il prossimo passo naturale è chiedersi quali curve realizzano effettivamente la distanza minima: le **geodetiche**, l'oggetto del prossimo notebook.
