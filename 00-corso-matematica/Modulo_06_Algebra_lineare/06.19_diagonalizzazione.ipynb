# 6.19 — Diagonalizzazione

**Prerequisiti:** 6.18 Autovettori, 6.16 Matrici invertibili, 6.14 Matrice associata.


## Perché ci interessa

Nel notebook 6.18 abbiamo visto che, con autovalori distinti, gli autovettori di una matrice $A$ formano una **base** di $\mathbb R^n$. Cosa succede se descriviamo l'applicazione lineare $T(x)=Ax$ usando **questa** base (fatta di autovettori) invece della base canonica? Per definizione di autovettore, $T$ agisce su ciascun vettore di base semplicemente moltiplicandolo per il relativo autovalore — nessuna "mescolanza" tra le coordinate. La matrice associata a $T$, rispetto a questa base speciale, diventa quindi **diagonale**: la forma più semplice possibile. Questo processo, chiamato **diagonalizzazione**, permette di capire immediatamente il comportamento di $A$ (per esempio calcolarne le potenze $A^k$, fondamentali nello studio dei sistemi dinamici del Modulo 8) senza fare calcoli complicati.

## Teoria, passo dopo passo

### La matrice diagonalizzante

Supponiamo che $A$ ($n\times n$) abbia $n$ autovettori indipendenti $v_1,\dots,v_n$, con autovalori (non necessariamente distinti) $\lambda_1,\dots,\lambda_n$. Costruiamo la matrice $P$ le cui colonne sono $v_1,\dots,v_n$ (invertibile, perché le colonne sono indipendenti, notebook 6.16). Allora:

$$AP = A\begin{pmatrix}v_1&\cdots&v_n\end{pmatrix} = \begin{pmatrix}Av_1&\cdots&Av_n\end{pmatrix} = \begin{pmatrix}\lambda_1v_1&\cdots&\lambda_nv_n\end{pmatrix} = PD$$

dove $D=\operatorname{diag}(\lambda_1,\dots,\lambda_n)$ è la matrice diagonale con gli autovalori sulla diagonale (l'ultima uguaglianza si verifica moltiplicando $P$ per $D$ colonna per colonna: la colonna $i$ di $PD$ è $P$ applicata alla colonna $i$ di $D$, cioè $\lambda_i$ volte la colonna $i$ di $P$, che è $v_i$).

Da $AP=PD$, moltiplicando a destra per $P^{-1}$ (che esiste, essendo $P$ invertibile):

$$A = PDP^{-1} \qquad\text{equivalentemente}\qquad D = P^{-1}AP$$

Questa è la **diagonalizzazione** di $A$: $A$ è simile a una matrice diagonale $D$, tramite la matrice $P$ le cui colonne sono gli autovettori.

### Quando una matrice è diagonalizzabile

Una matrice $A$ ($n\times n$) è **diagonalizzabile** se ammette $n$ autovettori linearmente indipendenti — equivalentemente, se esiste una base di $\mathbb R^n$ fatta di autovettori di $A$. Per il fatto visto nel notebook 6.18, **una condizione sufficiente** (ma non necessaria) è che $A$ abbia $n$ autovalori reali **distinti**: in tal caso $A$ è sempre diagonalizzabile. Se un autovalore ha **molteplicità** maggiore di 1 (è radice multipla del polinomio caratteristico), la matrice potrebbe non essere diagonalizzabile: bisogna verificare che l'autospazio corrispondente abbia dimensione pari alla molteplicità.

### A cosa serve: calcolo delle potenze

Se $A=PDP^{-1}$, allora $A^2=PDP^{-1}PDP^{-1}=PD(P^{-1}P)DP^{-1}=PDDP^{-1}=PD^2P^{-1}$ (usando $P^{-1}P=I$, notebook 6.16). Per induzione (notebook 0.20), $A^k=PD^kP^{-1}$ per ogni $k$. E calcolare $D^k$ è immediato: essendo diagonale, $D^k=\operatorname{diag}(\lambda_1^k,\dots,\lambda_n^k)$ (basta elevare ogni autovalore alla $k$-esima potenza) — enormemente più semplice che moltiplicare $A$ per sé stessa $k$ volte.

````python
import sympy as sp

A = sp.Matrix([[3, 1], [0, 2]])  # gli stessi autovalori 3, 2 del notebook 6.17-6.18

# Diagonalizzazione con SymPy
P, D = A.diagonalize()
print("P (colonne = autovettori) =\n", P)
print("\nD (autovalori sulla diagonale) =\n", D)

# Verifica: A = P D P^-1
print("\nP*D*P^-1 =\n", P * D * P.inv())
print("Coincide con A?", sp.simplify(P * D * P.inv() - A) == sp.zeros(2, 2))

# Calcolo di A^5 tramite diagonalizzazione, confrontato con il calcolo diretto
A5_diag = P * D**5 * P.inv()
A5_diretto = A**5
print("\nA^5 (via diagonalizzazione) =\n", A5_diag)
print("A^5 (calcolo diretto) =\n", A5_diretto)
print("Coincidono?", sp.simplify(A5_diag - A5_diretto) == sp.zeros(2, 2))
````

## Esercizio svolto e commentato

**Problema:** diagonalizza $A=\begin{pmatrix}1&2\\2&1\end{pmatrix}$ (autovalori e autovettori già trovati nel notebook 6.18: $\lambda_1=3,v_1=(1,1)$; $\lambda_2=-1,v_2=(1,-1)$) e usala per calcolare $A^3$.

**Ragionamento:**

**Passo 1 — costruiamo $P$**, affiancando gli autovettori come colonne: $P=\begin{pmatrix}1&1\\1&-1\end{pmatrix}$.

**Passo 2 — costruiamo $D$**, con gli autovalori sulla diagonale, **nello stesso ordine** delle colonne di $P$ (fondamentale: la colonna $i$ di $P$ deve corrispondere all'elemento $i$-esimo della diagonale di $D$): $D=\begin{pmatrix}3&0\\0&-1\end{pmatrix}$.

**Passo 3 — calcoliamo $P^{-1}$** con la formula $2\times2$ (notebook 6.16): $\det P=1\times(-1)-1\times1=-2$. $P^{-1}=\frac1{-2}\begin{pmatrix}-1&-1\\-1&1\end{pmatrix}=\begin{pmatrix}1/2&1/2\\1/2&-1/2\end{pmatrix}$.

**Passo 4 — calcoliamo $D^3$**: $D^3=\begin{pmatrix}27&0\\0&-1\end{pmatrix}$ (basta elevare ogni autovalore alla terza: $3^3=27$, $(-1)^3=-1$).

**Passo 5 — calcoliamo $A^3=PD^3P^{-1}$**: $PD^3=\begin{pmatrix}1&1\\1&-1\end{pmatrix}\begin{pmatrix}27&0\\0&-1\end{pmatrix}=\begin{pmatrix}27&-1\\27&1\end{pmatrix}$. Poi $(PD^3)P^{-1}=\begin{pmatrix}27&-1\\27&1\end{pmatrix}\begin{pmatrix}1/2&1/2\\1/2&-1/2\end{pmatrix}=\begin{pmatrix}27/2-1/2&27/2+1/2\\27/2+1/2&27/2-1/2\end{pmatrix}=\begin{pmatrix}13&14\\14&13\end{pmatrix}$.

**Conclusione:** $A^3=\begin{pmatrix}13&14\\14&13\end{pmatrix}$ (verificabile anche moltiplicando $A$ per sé stessa tre volte, molto più lentamente).

**Perché il calcolo tramite diagonalizzazione conviene davvero, nonostante i passaggi intermedi ($P$, $P^{-1}$, $D^3$) sembrino più numerosi:** qui, per $A^3$, il vantaggio non è enorme; ma per $A^{100}$, calcolare $D^{100}$ resta immediato (basta elevare $3$ e $-1$ alla centesima potenza), mentre moltiplicare $A$ per sé stessa cento volte richiederebbe 99 moltiplicazioni di matrici $2\times2$. Il costo di calcolare $P$, $P^{-1}$ e $D^k$ non dipende da $k$: si paga una volta sola, indipendentemente da quanto grande sia l'esponente.

````python
import sympy as sp

A = sp.Matrix([[1, 2], [2, 1]])
P, D = A.diagonalize()
print("P =\n", P, "\nD =\n", D)

A3_diag = P * D**3 * P.inv()
print("\nA^3 (via diagonalizzazione) =\n", A3_diag)
print("A^3 (calcolo diretto) =\n", A**3)
````

**3.** $\det(A-\lambda I)=\det\begin{pmatrix}-\lambda&1\\0&-\lambda\end{pmatrix}=\lambda^2$: l'unico autovalore è $\lambda=0$, con molteplicità 2 (radice doppia). L'autospazio: $A-0I=A=\begin{pmatrix}0&1\\0&0\end{pmatrix}$, sistema $y=0$ con $x$ libero: $E_0=\operatorname{span}((1,0))$, di dimensione **1**, non 2 come la molteplicità dell'autovalore. Non essendoci abbastanza autovettori indipendenti (solo 1, ne servirebbero 2), $A$ **non è diagonalizzabile**.

**4.** $\det A=\det(PDP^{-1})=\det P\cdot\det D\cdot\det(P^{-1})$ (proprietà del determinante del prodotto, applicata due volte). Poiché $\det(P^{-1})=1/\det P$ (dal notebook 6.16, prendendo il determinante di $PP^{-1}=I$: $\det P\cdot\det(P^{-1})=\det I=1$), otteniamo $\det A=\det P\cdot\det D\cdot\frac1{\det P}=\det D$. Poiché $D$ è diagonale, $\det D$ è il prodotto degli elementi diagonali, cioè il prodotto degli autovalori.

## Riepilogo finale

- Se $A$ ($n\times n$) ha $n$ autovettori indipendenti, è **diagonalizzabile**: $A=PDP^{-1}$, con $P$ formata dagli autovettori (colonne) e $D$ diagonale con i relativi autovalori.
- Condizione sufficiente per la diagonalizzabilità: $n$ autovalori reali distinti.
- Se un autovalore ha molteplicità maggiore di 1, bisogna verificare che il suo autospazio abbia la stessa dimensione: altrimenti la matrice **non** è diagonalizzabile.
- La diagonalizzazione rende immediato il calcolo di $A^k=PD^kP^{-1}$, con $D^k$ calcolato elevando semplicemente ogni autovalore alla $k$-esima potenza.
