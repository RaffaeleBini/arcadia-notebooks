# 6.8 — Basi

**Prerequisiti:** 6.6 Indipendenza lineare, 6.7 Generatori.


## Perché ci interessa

Nel notebook 6.7 abbiamo visto che, eliminando vettori ridondanti da un insieme di generatori, si arriva a un insieme che è sia generatore sia linearmente indipendente: "minimo ma sufficiente". Questo tipo di insieme è così importante da meritare un nome: **base**. Una base è, in un certo senso, il miglior "sistema di riferimento" per uno spazio vettoriale: ci permette di descrivere ogni vettore con un numero minimo di coordinate, in modo unico. È il concetto che rende possibile, per esempio, rappresentare un punto dello spazio con esattamente 3 numeri (in $\mathbb R^3$) e non uno di più o uno di meno.

## Teoria, passo dopo passo

### Definizione

Un insieme di vettori $\mathcal B=\{v_1,\dots,v_n\}$ di uno spazio vettoriale $V$ è una **base** di $V$ se:

1. $v_1,\dots,v_n$ **generano** $V$ (notebook 6.7): $\operatorname{span}(v_1,\dots,v_n)=V$;
2. $v_1,\dots,v_n$ sono **linearmente indipendenti** (notebook 6.6).

### La base canonica di $\mathbb R^n$

L'esempio più naturale: in $\mathbb R^n$, i vettori $e_1=(1,0,\dots,0)$, $e_2=(0,1,0,\dots,0)$, ..., $e_n=(0,\dots,0,1)$ (un $1$ nella posizione $i$-esima, zero altrove) formano la **base canonica**. Generano $\mathbb R^n$ (ogni vettore $(x_1,\dots,x_n)=x_1e_1+\cdots+x_ne_n$, banalmente) e sono indipendenti (se $\lambda_1e_1+\cdots+\lambda_ne_n=(\lambda_1,\dots,\lambda_n)=\mathbf 0$, allora ogni $\lambda_i=0$).

### Il fatto centrale: coordinate uniche rispetto a una base

Ecco perché le basi sono così utili: se $\mathcal B=\{v_1,\dots,v_n\}$ è una base di $V$, ogni vettore $u\in V$ si scrive in **uno e un solo modo** come combinazione lineare di $v_1,\dots,v_n$.

**Dimostrazione.** L'esistenza segue dal fatto che $v_1,\dots,v_n$ generano $V$ (punto 1 della definizione): esiste almeno una scrittura $u=\lambda_1v_1+\cdots+\lambda_nv_n$. Per l'**unicità**, supponiamo che esistano due scritture: $u=\lambda_1v_1+\cdots+\lambda_nv_n$ e anche $u=\mu_1v_1+\cdots+\mu_nv_n$. Sottraendo membro a membro (proprietà invariantiva, notebook 0.5):

$$\mathbf 0 = (\lambda_1-\mu_1)v_1+\cdots+(\lambda_n-\mu_n)v_n$$

Ma $v_1,\dots,v_n$ sono linearmente indipendenti (punto 2 della definizione): l'unica combinazione che dà $\mathbf 0$ è quella con tutti i coefficienti nulli (notebook 6.6). Quindi $\lambda_i-\mu_i=0$ per ogni $i$, cioè $\lambda_i=\mu_i$: le due scritture erano in realtà la stessa. $\blacksquare$

I coefficienti $\lambda_1,\dots,\lambda_n$ (unici, per il fatto appena dimostrato) si chiamano **coordinate** di $u$ rispetto alla base $\mathcal B$.

### Perché servono entrambe le condizioni

Se $v_1,\dots,v_n$ generano $V$ ma **non** sono indipendenti, un vettore può avere **più** scritture (come visto nel notebook 6.7, con $(1,1)=1\cdot(1,0)+1\cdot(0,1)=$ ...oppure, usando anche $(1,1)$ stesso come terzo generatore ridondante, in modi diversi): niente unicità. Se invece $v_1,\dots,v_n$ sono indipendenti ma **non** generano $V$, alcuni vettori di $V$ non hanno **nessuna** scrittura in termini di essi: niente esistenza. Servono entrambe le proprietà, insieme.

### Basi diverse per lo stesso spazio

Uno spazio vettoriale ha, in generale, **infinite** basi diverse. Per esempio, in $\mathbb R^2$, sia $\{(1,0),(0,1)\}$ sia $\{(1,1),(1,-1)\}$ sono basi (entrambe generano e sono indipendenti, come verificato nei notebook precedenti). Cambiare base cambia le coordinate di un vettore, ma non il vettore stesso — un'idea che riprenderemo nel notebook 6.14 (matrice associata a un'applicazione lineare).

````python
import sympy as sp

# Base canonica di R^3: verifica generazione e indipendenza
e1, e2, e3 = sp.Matrix([1,0,0]), sp.Matrix([0,1,0]), sp.Matrix([0,0,1])
u1, u2, u3, l1, l2, l3 = sp.symbols('u1 u2 u3 lambda1 lambda2 lambda3')

# Generazione: ogni (u1,u2,u3) si ottiene con lambda_i = u_i
u = sp.Matrix([u1, u2, u3])
combinazione = u1*e1 + u2*e2 + u3*e3
print("u1*e1+u2*e2+u3*e3 =", combinazione.T, " (coincide sempre con u)")

# Indipendenza: unica soluzione del sistema omogeneo è quella nulla
eq = l1*e1 + l2*e2 + l3*e3
sol = sp.solve([eq[0], eq[1], eq[2]], [l1, l2, l3])
print("Soluzione sistema omogeneo:", sol)

# Coordinate di u=(4,-1,2) rispetto alla base B={(1,1,0),(0,1,1),(1,0,1)}
v1, v2, v3 = sp.Matrix([1,1,0]), sp.Matrix([0,1,1]), sp.Matrix([1,0,1])
target = sp.Matrix([4, -1, 2])
eqB = l1*v1 + l2*v2 + l3*v3 - target
coordinate = sp.solve([eqB[0], eqB[1], eqB[2]], [l1, l2, l3])
print("\nCoordinate di (4,-1,2) rispetto a B:", coordinate)
````

## Esercizio svolto e commentato

**Problema:** verifica che $\mathcal B=\{(1,1),(1,-1)\}$ è una base di $\mathbb R^2$, poi trova le coordinate di $u=(5,1)$ rispetto a $\mathcal B$.

**Ragionamento:**

**Passo 1 — verifichiamo l'indipendenza lineare**: impostiamo $\lambda_1(1,1)+\lambda_2(1,-1)=(0,0)$, cioè $\begin{cases}\lambda_1+\lambda_2=0\\ \lambda_1-\lambda_2=0\end{cases}$. Sommando le due equazioni: $2\lambda_1=0$, quindi $\lambda_1=0$; sostituendo, $\lambda_2=0$. Unica soluzione nulla: **indipendenti**.

**Passo 2 — verifichiamo la generazione** (già fatto nell'esempio del notebook 6.5, lo ripetiamo qui in generale): per un $u=(u_1,u_2)$ arbitrario, $\lambda_1+\lambda_2=u_1$, $\lambda_1-\lambda_2=u_2$. Sommando: $\lambda_1=\frac{u_1+u_2}2$; sottraendo: $\lambda_2=\frac{u_1-u_2}2$. Soluzione sempre esistente: **generano** $\mathbb R^2$.

**Passo 3 — concludiamo che $\mathcal B$ è una base** (entrambe le condizioni verificate), quindi ogni vettore ha coordinate uniche rispetto a $\mathcal B$.

**Passo 4 — calcoliamo le coordinate di $u=(5,1)$** usando le formule appena trovate: $\lambda_1=\frac{5+1}2=3$, $\lambda_2=\frac{5-1}2=2$.

**Verifica:** $3(1,1)+2(1,-1)=(3+2,\,3-2)=(5,1)$. ✓ Le coordinate di $u$ rispetto a $\mathcal B$ sono $(3,2)$ — diverse dalle sue coordinate $(5,1)$ rispetto alla base canonica, pur rappresentando lo **stesso** vettore.

**Perché è utile aver ricavato le formule generali per $\lambda_1,\lambda_2$ (passo 2) invece di risolvere il sistema da capo ogni volta:** avendo le formule $\lambda_1=\frac{u_1+u_2}2$, $\lambda_2=\frac{u_1-u_2}2$ in funzione di un $u$ generico, il calcolo delle coordinate di **qualsiasi** vettore rispetto a questa base diventa immediato — una sostituzione, non un nuovo sistema da risolvere. È un'anticipazione dell'idea di "cambiamento di coordinate" che tratteremo più sistematicamente studiando le applicazioni lineari (notebook 6.13-6.14).

````python
import sympy as sp

l1, l2 = sp.symbols('lambda1 lambda2')
v1, v2 = sp.Matrix([1, 1]), sp.Matrix([1, -1])
u = sp.Matrix([5, 1])

eq = l1 * v1 + l2 * v2 - u
coordinate = sp.solve([eq[0], eq[1]], [l1, l2])
print("Coordinate di (5,1) rispetto a B={(1,1),(1,-1)}:", coordinate)
print("Verifica:", coordinate[l1] * v1 + coordinate[l2] * v2)
````

## Riepilogo finale

- Una **base** $\mathcal B$ di $V$ è un insieme di vettori che genera $V$ **e** è linearmente indipendente.
- Ogni vettore di $V$ si scrive in **modo unico** come combinazione lineare degli elementi di una base: i coefficienti sono le sue **coordinate** rispetto a quella base.
- Uno spazio vettoriale ha in generale infinite basi diverse: cambiando base, cambiano le coordinate di un vettore, ma non il vettore stesso.
- La **base canonica** di $\mathbb R^n$ (i vettori $e_1,\dots,e_n$) è la scelta più naturale, ma non l'unica possibile.
