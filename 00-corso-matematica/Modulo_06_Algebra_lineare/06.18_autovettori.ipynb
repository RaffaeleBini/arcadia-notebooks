# 6.18 — Autovettori

**Prerequisiti:** 6.17 Autovalori, 6.15 Nucleo e immagine.


## Perché ci interessa

Nel notebook 6.17 abbiamo trovato gli autovalori risolvendo il polinomio caratteristico, ma per ogni autovalore esiste tipicamente **più di un** autovettore associato (per esempio, se $v$ è autovettore, anche $2v$ lo è, con lo stesso autovalore). Vogliamo descrivere sistematicamente **tutti** gli autovettori associati a un dato autovalore: scopriremo che formano un sottospazio — l'**autospazio** — che possiamo calcolare con gli stessi strumenti già visti per il nucleo (notebook 6.15).

## Teoria, passo dopo passo

### L'autospazio

Dato un autovalore $\lambda$ di $A$, l'insieme di **tutti** i vettori $v$ (incluso, per convenzione algebrica, il vettore nullo) che soddisfano $Av=\lambda v$ si chiama **autospazio** relativo a $\lambda$, indicato $E_\lambda$:

$$E_\lambda = \{v\in\mathbb R^n : Av=\lambda v\} = \{v\in\mathbb R^n : (A-\lambda I)v=\mathbf 0\} = \ker(A-\lambda I)$$

Riconosciamo l'ultima espressione: è esattamente il **nucleo** della matrice $A-\lambda I$ (notebook 6.15)! Per questo $E_\lambda$ è automaticamente un sottospazio (i nuclei lo sono sempre): la somma di due autovettori relativi allo stesso $\lambda$ è ancora un autovettore relativo a $\lambda$ (o il vettore nullo), e così un multiplo scalare. Gli **autovettori** veri e propri sono gli elementi non nulli di $E_\lambda$ (il vettore nullo non conta come autovettore, per definizione, anche se appartiene formalmente al sottospazio $E_\lambda$).

### Come calcolare l'autospazio

Per ogni autovalore $\lambda$ trovato (risolvendo il polinomio caratteristico, notebook 6.17), si risolve il sistema omogeneo $(A-\lambda I)v=\mathbf 0$ con le tecniche del notebook 6.11: le soluzioni formano $E_\lambda$.

### Esempio guidato

Riprendiamo $A=\begin{pmatrix}3&1\\0&2\end{pmatrix}$ del notebook 6.17, con autovalori $\lambda_1=3,\lambda_2=2$.

**Per $\lambda_1=3$**: $A-3I=\begin{pmatrix}0&1\\0&-1\end{pmatrix}$. Il sistema $(A-3I)v=\mathbf 0$ dà: $0\cdot x+1\cdot y=0$ (e la seconda riga, $-y=0$, è la stessa equazione moltiplicata per $-1$). Quindi $y=0$, con $x$ libero: $E_3=\{(x,0):x\in\mathbb R\}=\operatorname{span}((1,0))$.

**Per $\lambda_2=2$**: $A-2I=\begin{pmatrix}1&1\\0&0\end{pmatrix}$. Il sistema dà $x+y=0$, cioè $x=-y$, con $y$ libero: $E_2=\operatorname{span}((-1,1))$.

**Verifica**: $A(1,0)=(3,0)=3(1,0)$ ✓ (autovalore 3). $A(-1,1)=(-3+1,\,0+2)=(-2,2)=2(-1,1)$ ✓ (autovalore 2).

### Autovettori relativi ad autovalori distinti sono indipendenti

> **Fatto.** Se $v_1,\dots,v_k$ sono autovettori associati ad autovalori $\lambda_1,\dots,\lambda_k$ **tutti distinti**, allora $v_1,\dots,v_k$ sono linearmente indipendenti (notebook 6.6).

Questo fatto (che non dimostriamo qui nel caso generale, ma verifichiamo nell'esempio) è centrale per il prossimo notebook: se una matrice $n\times n$ ha $n$ autovalori distinti, i suoi $n$ autovettori corrispondenti sono automaticamente indipendenti — quindi, per la scorciatoia del notebook 6.9, formano una **base** di $\mathbb R^n$, fatta interamente di autovettori.

````python
import sympy as sp

A = sp.Matrix([[3, 1], [0, 2]])
I = sp.eye(2)

# Autospazio per lambda = 3
M1 = A - 3 * I
print("A - 3I =\n", M1)
E3 = M1.nullspace()
print("Base di E_3:", [v.T for v in E3])

# Autospazio per lambda = 2
M2 = A - 2 * I
print("\nA - 2I =\n", M2)
E2 = M2.nullspace()
print("Base di E_2:", [v.T for v in E2])

# Verifica: A*v = lambda*v
v3 = E3[0]
v2 = E2[0]
print("\nA*v3 =", (A * v3).T, " | 3*v3 =", (3 * v3).T)
print("A*v2 =", (A * v2).T, " | 2*v2 =", (2 * v2).T)

# Indipendenza di v3 e v2
c1, c2 = sp.symbols('c1 c2')
eq = c1 * v3 + c2 * v2
sol = sp.solve([eq[0], eq[1]], [c1, c2])
print("\nv3, v2 indipendenti? (unica soluzione nulla)", sol)

# Confronto con eigenvects() di SymPy (dà autovalore, molteplicità, autovettori)
print("\nA.eigenvects():", A.eigenvects())
````

## Esercizio svolto e commentato

**Problema:** trova gli autovalori e gli autospazi di $A=\begin{pmatrix}1&2\\2&1\end{pmatrix}$ (già incontrata nell'esercizio 2 del notebook 6.17, con autovalori $3,-1$).

**Ragionamento:**

**Passo 1 — per $\lambda_1=3$, calcoliamo $A-3I$**: $\begin{pmatrix}1-3&2\\2&1-3\end{pmatrix}=\begin{pmatrix}-2&2\\2&-2\end{pmatrix}$.

**Passo 2 — risolviamo $(A-3I)v=\mathbf 0$**: $-2x+2y=0$ (e la seconda riga, $2x-2y=0$, è la stessa equazione moltiplicata per $-1$: ridondante, come atteso — se non lo fosse, avremmo sbagliato calcolo, perché $\lambda=3$ è per costruzione una radice del polinomio caratteristico, quindi $\det(A-3I)=0$, e un sistema $2\times2$ omogeneo con determinante nullo ha sempre righe proporzionali). Da $-2x+2y=0$: $x=y$. $E_3=\operatorname{span}((1,1))$.

**Passo 3 — per $\lambda_2=-1$, calcoliamo $A-(-1)I=A+I$**: $\begin{pmatrix}2&2\\2&2\end{pmatrix}$.

**Passo 4 — risolviamo**: $2x+2y=0$, cioè $x=-y$. $E_{-1}=\operatorname{span}((1,-1))$.

**Passo 5 — verifichiamo l'indipendenza** dei due autovettori trovati, $(1,1)$ e $(1,-1)$: non sono multipli l'uno dell'altro (il rapporto tra le componenti è diverso: $1/1=1$ contro $1/(-1)=-1$), quindi sono indipendenti — coerente col fatto generale visto nella teoria (autovalori distinti $\Rightarrow$ autovettori indipendenti).

**Conclusione:** $E_3=\operatorname{span}((1,1))$, $E_{-1}=\operatorname{span}((1,-1))$; insieme, $(1,1)$ e $(1,-1)$ formano una base di $\mathbb R^2$ fatta di autovettori.

**Perché la ridondanza tra le righe del sistema (passo 2) non è mai un caso, quando $\lambda$ è davvero un autovalore:** per costruzione, $\lambda$ è scelto in modo che $\det(A-\lambda I)=0$ (equazione caratteristica, notebook 6.17); per una matrice $2\times2$, determinante nullo significa righe (o colonne) proporzionali (notebook 6.12), quindi il sistema $2\times2$ ha sempre almeno un'equazione ridondante — è il motivo strutturale per cui l'autospazio contiene sempre più del solo vettore nullo (altrimenti $\lambda$ non sarebbe stato un autovalore, per definizione).

````python
import sympy as sp

A = sp.Matrix([[1, 2], [2, 1]])

print("A.eigenvects():")
for autoval, molt, autovett in A.eigenvects():
    print(f"  lambda={autoval}, molteplicità={molt}, autovettori={[v.T for v in autovett]}")
````

## Riepilogo finale

- L'**autospazio** $E_\lambda=\ker(A-\lambda I)$ è il sottospazio di tutti gli autovettori relativi a $\lambda$ (più il vettore nullo); si calcola risolvendo il sistema omogeneo $(A-\lambda I)v=\mathbf 0$.
- Autovettori relativi ad autovalori **distinti** sono sempre linearmente indipendenti.
- Se una matrice $n\times n$ ha $n$ autovalori distinti, i corrispondenti autovettori formano automaticamente una **base** di $\mathbb R^n$.
- Questo fatto apre la strada al prossimo notebook: usare una base di autovettori per semplificare drasticamente la struttura di una matrice.
