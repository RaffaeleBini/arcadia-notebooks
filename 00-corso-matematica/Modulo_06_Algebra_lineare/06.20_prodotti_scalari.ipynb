## Teoria, passo dopo passo

### Cosa rendeva speciale il prodotto scalare di $\mathbb R^n$?

Ripercorriamo le proprietà del prodotto scalare $v\cdot w=\sum v_iw_i$ visto nel notebook 6.2, che ci servirono per definire norma e angolo:

- **simmetria**: $v\cdot w=w\cdot v$;
- **bilinearità**: $v\cdot(w+u)=v\cdot w+v\cdot u$ e $(\lambda v)\cdot w=\lambda(v\cdot w)$ (lineare in ciascun argomento);
- **positività**: $v\cdot v\ge0$, con uguaglianza solo per $v=\mathbf 0$.

### Definizione astratta di prodotto scalare

Uno **prodotto scalare** (o prodotto interno) su uno spazio vettoriale $V$ è una funzione $\langle\cdot,\cdot\rangle:V\times V\to\mathbb R$ che soddisfa esattamente queste tre proprietà, per ogni $u,v,w\in V$ e $\lambda\in\mathbb R$:

1. **simmetria**: $\langle v,w\rangle=\langle w,v\rangle$;
2. **bilinearità**: $\langle v,w+u\rangle=\langle v,w\rangle+\langle v,u\rangle$ e $\langle\lambda v,w\rangle=\lambda\langle v,w\rangle$;
3. **positività (definita)**: $\langle v,v\rangle\ge0$, con $\langle v,v\rangle=0$ se e solo se $v=\mathbf 0$.

Uno spazio vettoriale dotato di un prodotto scalare si chiama **spazio euclideo** (in dimensione finita).

### Norma indotta

Esattamente come in $\mathbb R^n$, definiamo la **norma** $\|v\|=\sqrt{\langle v,v\rangle}$ (ben definita: $\langle v,v\rangle\ge0$ per la positività, quindi la radice esiste). Tutte le proprietà geometriche viste nel notebook 6.2 — la disuguaglianza di Cauchy-Schwarz, la definizione di angolo, l'ortogonalità — si estendono automaticamente a **qualsiasi** spazio con prodotto scalare, senza bisogno di ridimostrarle da zero: dipendono solo dalle tre proprietà astratte, non dalla forma specifica di $V$.

### Esempio: prodotto scalare tra polinomi

Su $\mathbb R^2[x]$ (polinomi di grado $\le2$, notebook 6.3), definiamo:

$$\langle p,q\rangle = \int_0^1 p(x)q(x)\,dx$$

(l'integrale sarà formalizzato nel Modulo 5, ma qui basta pensarlo come "l'area sotto il prodotto $p(x)q(x)$" — un numero ben definito per ogni coppia di polinomi). Si può verificare che soddisfa le tre proprietà: la simmetria è ovvia ($p(x)q(x)=q(x)p(x)$); la bilinearità segue dalla linearità dell'integrale; la positività segue dal fatto che $p(x)^2\ge0$ ovunque, quindi il suo integrale è $\ge0$, ed è $0$ solo se $p(x)^2$ è identicamente nullo, cioè $p=\mathbf 0$.

Con questo prodotto scalare, $\|1\|=\sqrt{\int_0^1 1\,dx}=\sqrt1=1$, mentre $\|x\|=\sqrt{\int_0^1x^2dx}=\sqrt{1/3}$: i "vettori" $1$ e $x$ hanno lunghezze diverse, un'idea non ovvia a priori ma perfettamente coerente con la struttura astratta.

````python
import sympy as sp

x = sp.symbols('x')

def prodotto_scalare_polinomi(p, q):
    return sp.integrate(p * q, (x, 0, 1))

p1, p2, p3 = sp.Integer(1), x, x**2

print("<1,1> =", prodotto_scalare_polinomi(p1, p1))
print("<x,x> =", prodotto_scalare_polinomi(p2, p2))
print("<1,x> =", prodotto_scalare_polinomi(p1, p2))

# Norme indotte
norma_1 = sp.sqrt(prodotto_scalare_polinomi(p1, p1))
norma_x = sp.sqrt(prodotto_scalare_polinomi(p2, p2))
print("\n||1|| =", norma_1)
print("||x|| =", norma_x)

# Verifica simmetria
print("\n<1,x> == <x,1>?", prodotto_scalare_polinomi(p1, p2) == prodotto_scalare_polinomi(p2, p1))

# Verifica bilinearità: <1, x + x^2> == <1,x> + <1,x^2>
lhs = prodotto_scalare_polinomi(p1, p2 + p3)
rhs = prodotto_scalare_polinomi(p1, p2) + prodotto_scalare_polinomi(p1, p3)
print("Bilinearità:", lhs, "==", rhs, "?", lhs == rhs)
````

## Esercizio svolto e commentato

**Problema:** usando il prodotto scalare $\langle p,q\rangle=\int_0^1p(x)q(x)dx$, stabilisci se $1$ e $x-\frac12$ sono ortogonali.

**Ragionamento:**

**Passo 1 — applichiamo il criterio di ortogonalità visto per $\mathbb R^n$** (notebook 6.2), che vale in ogni spazio con prodotto scalare (proprio perché dipende solo dalle tre proprietà astratte, non dalla natura specifica dei "vettori"): $p\perp q \iff \langle p,q\rangle=0$.

**Passo 2 — calcoliamo $\langle 1,\,x-\frac12\rangle$**: per bilinearità, $\langle1,x-\frac12\rangle=\langle1,x\rangle-\frac12\langle1,1\rangle$.

**Passo 3 — calcoliamo i due integrali necessari**: $\langle1,x\rangle=\int_0^1x\,dx=\left[\frac{x^2}2\right]_0^1=\frac12$. $\langle1,1\rangle=\int_0^11\,dx=1$.

**Passo 4 — componiamo**: $\langle1,x-\frac12\rangle=\frac12-\frac12\times1=\frac12-\frac12=0$.

**Conclusione:** sì, $1$ e $x-\frac12$ sono **ortogonali** rispetto a questo prodotto scalare.

**Perché questo risultato non è un caso isolato, ma la base della "correzione" $x-\frac12$:** il polinomio $x-\frac12$ è ottenuto sottraendo da $x$ la sua **proiezione** sul sottospazio generato da $1$ (esattamente come nel notebook 6.2: $\operatorname{proj}_1x=\frac{\langle x,1\rangle}{\langle1,1\rangle}\times1=\frac{1/2}1=\frac12$). Questa operazione — sottrarre la proiezione per ottenere un vettore ortogonale — è il cuore del procedimento di **ortogonalizzazione**, che formalizzeremo sistematicamente nel notebook 6.21.

````python
import sympy as sp

x = sp.symbols('x')

def prod(p, q):
    return sp.integrate(p * q, (x, 0, 1))

p, q = sp.Integer(1), x - sp.Rational(1, 2)
print("<1, x-1/2> =", prod(p, q))
print("Ortogonali?", prod(p, q) == 0)

proiezione = prod(x, 1) / prod(sp.Integer(1), sp.Integer(1))
print("\nProiezione di x su 1:", proiezione)
print("x - proiezione =", x - proiezione)
````

## Riepilogo finale

- Un **prodotto scalare** astratto su $V$ è una funzione simmetrica, bilineare e definita positiva: generalizza il prodotto scalare di $\mathbb R^n$ (notebook 6.2) a spazi vettoriali qualunque.
- Ogni prodotto scalare induce una **norma** $\|v\|=\sqrt{\langle v,v\rangle}$, e con essa tutte le nozioni geometriche (angolo, ortogonalità) del notebook 6.2, che valgono in generale perché dipendono solo dalle tre proprietà astratte.
- Esempi non banali: prodotti scalari "pesati" su $\mathbb R^n$, o l'integrale del prodotto tra funzioni/polinomi.
- Cambiare prodotto scalare cambia effettivamente lunghezze e angoli, pur restando nello stesso spazio vettoriale.
