## Perché ci interessa

Nel notebook 2.5-2.7 abbiamo studiato parabole, ellissi e iperboli tramite equazioni come $x^2+y^2=1$ (circonferenza) o $\frac{x^2}{a^2}+\frac{y^2}{b^2}=1$ (ellisse). Espressioni come $x^2+y^2$, o più in generale $ax^2+bxy+cy^2$, si chiamano **forme quadratiche**: appaiono ovunque — nell'equazione delle coniche, nell'energia cinetica in fisica, nella funzione di costo dei minimi quadrati in statistica, nei test di massimo/minimo per funzioni di più variabili (Modulo 7). L'algebra lineare, tramite autovalori e matrici ortogonali, ci dà uno strumento potentissimo per classificarle e semplificarle.

## Teoria, passo dopo passo

### Definizione

Una **forma quadratica** su $\mathbb R^n$ è una funzione del tipo:

$$q(v) = v^TAv$$

dove $A$ è una matrice **simmetrica** ($A=A^T$, cioè $a_{ij}=a_{ji}$) $n\times n$. Per $n=2$, con $A=\begin{pmatrix}a&b\\b&c\end{pmatrix}$ (simmetrica: l'elemento fuori diagonale è ripetuto), esplicitando $v=(x,y)$:

$$q(x,y) = v^TAv = ax^2+2bxy+cy^2$$

**Perché si richiede $A$ simmetrica:** ogni forma quadratica $ax^2+bxy+cy^2$ ammette **infinite** rappresentazioni come $v^TMv$ per matrici $M$ non simmetriche (il termine misto $xy$ si può "distribuire" in vari modi tra le posizioni fuori diagonale), ma **una sola** rappresentazione con $M$ simmetrica: prendendo la matrice simmetrica, evitiamo ambiguità, e — come vedremo — possiamo sfruttare i risultati sulla diagonalizzazione (notebook 6.19), che per le matrici simmetriche hanno una forma particolarmente potente.

### Il teorema spettrale (anticipazione)

Il prossimo notebook dimostrerà che **ogni matrice simmetrica reale è diagonalizzabile con una matrice ortogonale**: $A=QDQ^T$, con $Q$ ortogonale (notebook 6.22) e $D$ diagonale con gli autovalori reali di $A$. Anticipiamo qui la conseguenza pratica: sostituendo $v=Qw$ (un cambiamento di coordinate, con $Q$ ortogonale — quindi che non deforma lunghezze né angoli, notebook 6.22):

$$q(v) = v^TAv = (Qw)^TA(Qw) = w^TQ^TAQw = w^TQ^TQDQ^TQw = w^TDw = \lambda_1w_1^2+\cdots+\lambda_nw_n^2$$

(usando $Q^TQ=I$ due volte). In queste nuove coordinate $w$ (ottenute ruotando/riflettendo, senza deformare), la forma quadratica diventa una **somma di quadrati pesati dagli autovalori**, senza più termini misti: la forma più semplice possibile.

### Classificazione tramite il segno degli autovalori

Grazie a questa riduzione, il comportamento di $q(v)=\lambda_1w_1^2+\cdots+\lambda_nw_n^2$ dipende solo dai **segni** degli autovalori (poiché ogni $w_i^2\ge0$):

- **definita positiva**: tutti gli autovalori $>0$ ($q(v)>0$ per ogni $v\ne\mathbf 0$) — geometricamente, un'ellisse (o ellissoide);
- **definita negativa**: tutti $<0$;
- **indefinita**: autovalori di segno misto — geometricamente, un'iperbole;
- **semidefinita**: tutti $\ge0$ (o $\le0$), con almeno un autovalore nullo — geometricamente, una parabola o un caso degenere.

### Un test rapido per $2\times2$: il discriminante

Per $A=\begin{pmatrix}a&b\\b&c\end{pmatrix}$, senza calcolare esplicitamente gli autovalori, si può usare $\det A=ac-b^2$ e $\operatorname{tr}(A)=a+c$: se $\det A>0$ e $\operatorname{tr}(A)>0$, entrambi gli autovalori sono positivi (definita positiva); se $\det A>0$ e $\operatorname{tr}(A)<0$, entrambi negativi; se $\det A<0$, gli autovalori hanno segno opposto (indefinita) — perché $\det A$ è il **prodotto** degli autovalori (notebook 6.17), e $\operatorname{tr}(A)$ la loro somma.

````python
import sympy as sp

x, y = sp.symbols('x y', real=True)

# Forma quadratica x^2 + y^2 (circonferenza/ellisse): A = I
A1 = sp.Matrix([[1, 0], [0, 1]])
v = sp.Matrix([x, y])
q1 = (v.T * A1 * v)[0]
print("q1(x,y) =", sp.expand(q1))
print("Autovalori di A1:", A1.eigenvals())

# Forma quadratica x^2 - y^2 (iperbole): A = diag(1,-1)
A2 = sp.Matrix([[1, 0], [0, -1]])
q2 = (v.T * A2 * v)[0]
print("\nq2(x,y) =", sp.expand(q2))
print("Autovalori di A2:", A2.eigenvals())

# Forma quadratica con termine misto: 2x^2 + 2xy + 2y^2 -> A = [[2,1],[1,2]]
A3 = sp.Matrix([[2, 1], [1, 2]])
q3 = (v.T * A3 * v)[0]
print("\nq3(x,y) =", sp.expand(q3))
print("Autovalori di A3:", A3.eigenvals())
print("det(A3) =", A3.det(), " tr(A3) =", A3.trace(), "-> definita positiva se entrambi > 0")
````

## Esercizio svolto e commentato

**Problema:** classifica la forma quadratica $q(x,y)=5x^2-4xy+2y^2$.

**Ragionamento:**

**Passo 1 — costruiamo la matrice simmetrica associata.** Confrontando $q(x,y)=ax^2+2bxy+cy^2$ con $5x^2-4xy+2y^2$: $a=5$, $2b=-4$ quindi $b=-2$, $c=2$. La matrice è $A=\begin{pmatrix}5&-2\\-2&2\end{pmatrix}$.

**Passo 2 — usiamo il test rapido invece di calcolare esplicitamente gli autovalori**: $\det A=5\times2-(-2)\times(-2)=10-4=6$. $\operatorname{tr}(A)=5+2=7$.

**Passo 3 — interpretiamo**: $\det A=6>0$ (autovalori con lo stesso segno, essendo il loro prodotto positivo) e $\operatorname{tr}(A)=7>0$ (la loro somma è positiva, quindi — avendo lo stesso segno — devono essere entrambi positivi, non entrambi negativi).

**Conclusione:** la forma quadratica è **definita positiva**: $q(x,y)>0$ per ogni $(x,y)\ne(0,0)$. Geometricamente, l'insieme $\{q(x,y)=1\}$ è un'ellisse.

**Verifica diretta (opzionale, per controllo)**: calcoliamo esplicitamente gli autovalori con l'equazione caratteristica $\lambda^2-7\lambda+6=0$, cioè $\lambda=\frac{7\pm\sqrt{49-24}}2=\frac{7\pm5}2$, cioè $\lambda_1=6,\lambda_2=1$: entrambi positivi, confermando la classificazione.

**Perché il test rapido (determinante e traccia) è sufficiente in dimensione 2, senza dover risolvere esplicitamente il polinomio caratteristico:** in $2\times2$, conoscere **prodotto** e **somma** di due numeri reali basta per determinarne il segno collettivo (senza conoscerli singolarmente) — esattamente come, per un'equazione di secondo grado (notebook 1.6), le relazioni tra i coefficienti e le radici (somma $=-b/a$, prodotto $=c/a$) permettono deduzioni sui segni delle radici senza risolvere esplicitamente l'equazione. Questa scorciatoia, però, funziona comodamente solo in dimensione 2; per $n\times n$ più grande, servono criteri più elaborati (come quello dei minori principali, non trattato qui).

````python
import sympy as sp

A = sp.Matrix([[5, -2], [-2, 2]])
print("det(A) =", A.det(), " tr(A) =", A.trace())
print("Autovalori:", A.eigenvals())

x, y = sp.symbols('x y', real=True)
v = sp.Matrix([x, y])
q = (v.T * A * v)[0]
print("q(x,y) =", sp.expand(q))
````

## Riepilogo finale

- Una **forma quadratica** $q(v)=v^TAv$ è associata a un'unica matrice **simmetrica** $A$.
- Cambiando coordinate con una matrice ortogonale (che diagonalizza $A$), $q$ diventa una somma di quadrati pesati dagli autovalori, senza termini misti.
- Il **segno degli autovalori** classifica $q$ come definita positiva/negativa, indefinita, o semidefinita — con interpretazione geometrica come ellisse, iperbole, o parabola.
- In dimensione 2, il test rapido con $\det A$ e $\operatorname{tr}(A)$ evita di risolvere esplicitamente l'equazione caratteristica.
