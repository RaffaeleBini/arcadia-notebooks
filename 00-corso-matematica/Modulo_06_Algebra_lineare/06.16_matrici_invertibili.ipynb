# 6.16 — Matrici invertibili

**Prerequisiti:** 6.15 Nucleo e immagine, 6.12 Determinanti, 3.6 Funzione inversa.


## Perché ci interessa

Nel notebook 3.6 abbiamo visto che una funzione biiettiva ammette un'inversa, che "annulla" l'effetto della funzione originale. Per un'applicazione lineare invertibile $T$, la sua inversa $T^{-1}$ è ancora lineare (lo verificheremo), e — grazie alla corrispondenza tra applicazioni lineari e matrici (notebook 6.14) — corrisponde a una **matrice inversa** $A^{-1}$. Sapere quando una matrice è invertibile, e come calcolarne l'inversa, è centrale per risolvere sistemi lineari "in blocco" (per esempio con $b$ diversi ma la stessa $A$) e per capire quando una trasformazione geometrica può essere "annullata".

## Teoria, passo dopo passo

### Definizione

Una matrice quadrata $A$ ($n\times n$) è **invertibile** se esiste una matrice $A^{-1}$ ($n\times n$) tale che:

$$AA^{-1}=A^{-1}A=I$$

dove $I$ è la **matrice identità** (con $1$ sulla diagonale, $0$ altrove): la matrice associata all'applicazione identica $T(v)=v$, che lascia invariato ogni vettore.

### Caratterizzazione: quando una matrice è invertibile

Mettendo insieme i risultati dei notebook precedenti, per una matrice quadrata $A$ ($n\times n$) le seguenti affermazioni sono **tutte equivalenti** (ciascuna implica tutte le altre):

1. $A$ è invertibile;
2. $\det A\ne0$ (notebook 6.12);
3. le colonne di $A$ sono linearmente indipendenti (notebook 6.6);
4. le colonne di $A$ formano una base di $\mathbb R^n$ (notebook 6.8, per la scorciatoia del notebook 6.9, essendo $n$ vettori);
5. $Ax=b$ ha soluzione unica per ogni $b\in\mathbb R^n$ (notebook 6.11);
6. $\ker T=\{\mathbf 0\}$, dove $T(x)=Ax$ (notebook 6.15): $T$ è iniettiva;
7. $\operatorname{Im}T=\mathbb R^n$: $T$ è suriettiva (equivalente a 6, per il teorema della dimensione con $\dim V=\dim W=n$, notebook 6.15).

**Perché questa catena di equivalenze ha senso**: ciascuna riformula la stessa idea sotto un aspetto diverso — "$A$ non fa collassare nessuna informazione" — vista rispettivamente come proprietà algebrica del determinante, come indipendenza delle colonne, come esistenza/unicità delle soluzioni, o come iniettività/suriettività dell'applicazione associata. È uno dei risultati più utili dell'algebra lineare elementare, perché permette di verificare l'invertibilità nel modo più comodo caso per caso (spesso, il determinante è il test più rapido).

### Formula per l'inversa $2\times2$

Per $A=\begin{pmatrix}a&b\\c&d\end{pmatrix}$ con $\det A=ad-bc\ne0$:

$$A^{-1} = \frac1{\det A}\begin{pmatrix}d&-b\\-c&a\end{pmatrix}$$

**Verifica diretta:** $AA^{-1}=\frac1{\det A}\begin{pmatrix}a&b\\c&d\end{pmatrix}\begin{pmatrix}d&-b\\-c&a\end{pmatrix}=\frac1{\det A}\begin{pmatrix}ad-bc&-ab+ab\\cd-dc&-cb+ad\end{pmatrix}=\frac1{\det A}\begin{pmatrix}\det A&0\\0&\det A\end{pmatrix}=I$.

### Risolvere $Ax=b$ con l'inversa

Se $A$ è invertibile, l'equazione $Ax=b$ si risolve moltiplicando entrambi i membri a sinistra per $A^{-1}$ (proprietà invariantiva, notebook 0.5, adattata al prodotto matriciale, che però va applicato dallo stesso lato su entrambi i membri, essendo non commutativo, notebook 6.10): $A^{-1}Ax=A^{-1}b$, cioè $Ix=A^{-1}b$, cioè $x=A^{-1}b$. In pratica, per un singolo sistema, l'eliminazione di Gauss (notebook 6.11) resta più efficiente; calcolare $A^{-1}$ conviene quando si devono risolvere **molti** sistemi con la stessa $A$ e $b$ diversi.

````python
import sympy as sp

A = sp.Matrix([[2, 1], [1, 1]])
print("det(A) =", A.det())

A_inv = A.inv()
print("\nA^-1 =\n", A_inv)

I = sp.eye(2)
print("\nA * A^-1 =\n", A * A_inv, " -> uguale a I?", A * A_inv == I)

# Formula esplicita 2x2, confrontata con sp.inv()
a, b, c, d = A[0,0], A[0,1], A[1,0], A[1,1]
formula = (1 / A.det()) * sp.Matrix([[d, -b], [-c, a]])
print("\nFormula esplicita:\n", formula)
print("Coincide con A.inv()?", formula == A_inv)

# Risoluzione di Ax=b tramite l'inversa
b_vett = sp.Matrix([5, 3])
x = A_inv * b_vett
print("\nSoluzione x = A^-1 b:", x.T)
print("Verifica A@x:", (A * x).T)
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $A=\begin{pmatrix}1&2&3\\0&1&4\\5&6&0\end{pmatrix}$ è invertibile; se sì, usa $A^{-1}$ per risolvere $Ax=(1,0,0)$.

**Ragionamento:**

**Passo 1 — calcoliamo il determinante** (criterio più rapido, punto 2 dell'elenco), sviluppando lungo la prima colonna: $\det A=1\cdot\det\begin{pmatrix}1&4\\6&0\end{pmatrix}-0\cdot(\dots)+5\cdot\det\begin{pmatrix}2&3\\1&4\end{pmatrix}$.

**Passo 2 — calcoliamo i due minori necessari**: $\det\begin{pmatrix}1&4\\6&0\end{pmatrix}=0-24=-24$. $\det\begin{pmatrix}2&3\\1&4\end{pmatrix}=8-3=5$.

**Passo 3 — componiamo il risultato** (segni alternati $+,-,+$ lungo la colonna, con il termine centrale nullo per il coefficiente $0$ della matrice): $\det A=1\times(-24)-0+5\times5=-24+25=1$.

**Passo 4 — interpretiamo**: $\det A=1\ne0$, quindi $A$ **è invertibile**.

**Passo 5 — invece di calcolare $A^{-1}$ interamente a mano (procedimento lungo per una $3\times3$), usiamo SymPy** per ottenere $A^{-1}$ e poi calcolare $x=A^{-1}b$ con $b=(1,0,0)$.

**Conclusione:** con $\det A=1$ (un caso particolarmente comodo, perché $A^{-1}$ avrà tutte le entrate intere, essendo $\frac1{\det A}=1$), il sistema ha soluzione unica, calcolabile direttamente moltiplicando $A^{-1}$ per $b$.

**Perché conviene sempre verificare prima il determinante, prima di lanciarsi nel calcolo dell'inversa:** calcolare l'inversa di una matrice $3\times3$ o più grande è un procedimento lungo (richiede di calcolare $n^2$ minori, uno per ogni entrata della cosiddetta matrice aggiunta); verificare $\det A\ne0$ con lo sviluppo di Laplace è molto più rapido, ed evita di scoprire — dopo un lungo calcolo — che la matrice non era invertibile.

````python
import sympy as sp

A = sp.Matrix([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
print("det(A) =", A.det())

A_inv = A.inv()
print("\nA^-1 =\n", A_inv)

b = sp.Matrix([1, 0, 0])
x = A_inv * b
print("\nSoluzione x =", x.T)
print("Verifica A@x:", (A * x).T)
````

**4.** Vogliamo mostrare che $B^{-1}A^{-1}$ è l'inversa di $AB$, cioè che $(AB)(B^{-1}A^{-1})=I$: $(AB)(B^{-1}A^{-1})=A(BB^{-1})A^{-1}=AIA^{-1}=AA^{-1}=I$ (usando l'associatività del prodotto matriciale, notebook 6.10, per raggruppare $BB^{-1}=I$ prima). ## Riepilogo finale

- Una matrice quadrata $A$ è **invertibile** se esiste $A^{-1}$ con $AA^{-1}=A^{-1}A=I$.
- L'invertibilità è equivalente a: $\det A\ne0$, colonne indipendenti, colonne che formano una base, $Ax=b$ risolubile in modo unico per ogni $b$, $\ker T=\{\mathbf 0\}$, $\operatorname{Im}T=\mathbb R^n$ — sette modi diversi di dire la stessa cosa.
- Per $2\times2$: $A^{-1}=\frac1{\det A}\begin{pmatrix}d&-b\\-c&a\end{pmatrix}$.
- $(AB)^{-1}=B^{-1}A^{-1}$: attenzione all'ordine invertito, conseguenza della non commutatività del prodotto matriciale.
