# 6.6 — Indipendenza lineare

**Prerequisiti:** 6.5 Combinazioni lineari.


## Perché ci interessa

Nel notebook 6.5 abbiamo visto che $(2,2)$ non aggiunge nulla di nuovo allo span di $(1,1)$: è "ridondante", perché è già una combinazione lineare (un multiplo) di $(1,1)$. Questa ridondanza è un problema quando vogliamo descrivere uno spazio nel modo più **efficiente** possibile: se sto scegliendo un insieme minimo di "ingredienti" per generare un intero spazio (o sottospazio), non voglio includere vettori superflui. Il concetto di **indipendenza lineare** cattura esattamente l'assenza di questo tipo di ridondanza, ed è il passo che ci separa dal concetto di **base**, che vedremo tra due notebook.

## Teoria, passo dopo passo

### L'idea: nessun vettore è "spiegabile" dagli altri

Diciamo che $v_1,\dots,v_k$ sono **linearmente dipendenti** se almeno uno di essi è combinazione lineare degli altri — cioè è ridondante, "spiegabile" a partire dal resto. Al contrario, sono **linearmente indipendenti** se nessuno di essi si può scrivere come combinazione lineare degli altri: ognuno porta un contributo "nuovo" allo span.

### Definizione operativa (equazione omogenea)

Questa idea intuitiva si traduce in un criterio preciso e facile da verificare. I vettori $v_1,\dots,v_k$ sono **linearmente indipendenti** se l'unica combinazione lineare che dà il vettore nullo è quella con **tutti** i coefficienti nulli:

$$\lambda_1v_1+\lambda_2v_2+\cdots+\lambda_kv_k=\mathbf 0 \quad\Longrightarrow\quad \lambda_1=\lambda_2=\cdots=\lambda_k=0$$

Se invece esiste una combinazione con **non tutti** i coefficienti nulli che dà $\mathbf 0$ (detta combinazione **non banale**), i vettori sono **linearmente dipendenti**.

**Perché questa definizione cattura l'idea di "nessuna ridondanza":** supponiamo che $v_1,\dots,v_k$ siano dipendenti secondo questa definizione, cioè esista una combinazione non banale $\lambda_1v_1+\cdots+\lambda_kv_k=\mathbf 0$ con, diciamo, $\lambda_1\ne0$. Allora possiamo isolare $v_1$ (dividendo per $\lambda_1$, lecito perché $\lambda_1\ne0$, e usando la proprietà invariantiva, notebook 0.5):

$$v_1 = -\frac{\lambda_2}{\lambda_1}v_2-\cdots-\frac{\lambda_k}{\lambda_1}v_k$$

cioè $v_1$ è combinazione lineare degli altri: è proprio la ridondanza che volevamo formalizzare. Viceversa, se un $v_i$ è combinazione lineare degli altri, portando tutto a sinistra dell'uguale si ottiene una combinazione non banale (il coefficiente di $v_i$ è $-1\ne0$) che dà $\mathbf 0$.

### Esempi

- $v_1=(1,0)$, $v_2=(0,1)$: se $\lambda_1(1,0)+\lambda_2(0,1)=(0,0)$, allora $(\lambda_1,\lambda_2)=(0,0)$, cioè $\lambda_1=\lambda_2=0$: **indipendenti**.
- $v_1=(1,1)$, $v_2=(2,2)$: la combinazione $2v_1-v_2=(2,2)-(2,2)=(0,0)$ è non banale (coefficienti $2,-1$, non entrambi nulli): **dipendenti**. Coerente con l'osservazione del notebook 6.5.
- **Un vettore nullo rende sempre l'insieme dipendente**: se uno dei $v_i=\mathbf 0$, allora la combinazione con $\lambda_i=1$ e tutti gli altri coefficienti nulli dà $\mathbf 0$, ed è non banale.
- **Un solo vettore $v\ne\mathbf 0$ è sempre indipendente** (visto isolatamente): $\lambda v=\mathbf 0$ con $v\ne\mathbf 0$ implica $\lambda=0$ (se fosse $\lambda\ne0$, potremmo dividere per $\lambda$ ottenendo $v=\mathbf 0$, contraddizione).

### Verifica pratica: un sistema lineare omogeneo

Come per lo span, verificare l'indipendenza lineare di $v_1,\dots,v_k\in\mathbb R^n$ si riduce a risolvere il **sistema lineare omogeneo** $\lambda_1v_1+\cdots+\lambda_kv_k=\mathbf 0$ (omogeneo perché il termine noto è il vettore nullo, notebook 1.7). Se l'unica soluzione è quella nulla, i vettori sono indipendenti; se ci sono altre soluzioni, sono dipendenti.

````python
import sympy as sp

# v1=(1,0), v2=(0,1): indipendenti?
l1, l2 = sp.symbols('lambda1 lambda2')
v1, v2 = sp.Matrix([1, 0]), sp.Matrix([0, 1])
eq = l1 * v1 + l2 * v2
sol = sp.solve([eq[0], eq[1]], [l1, l2])
print("v1=(1,0), v2=(0,1) -> soluzione sistema omogeneo:", sol)

# v1=(1,1), v2=(2,2): dipendenti?
w1, w2 = sp.Matrix([1, 1]), sp.Matrix([2, 2])
eq2 = l1 * w1 + l2 * w2
sol2 = sp.solve([eq2[0], eq2[1]], [l1, l2])
print("v1=(1,1), v2=(2,2) -> soluzione sistema omogeneo:", sol2)
# se la soluzione ha un parametro libero, ci sono infinite combinazioni non banali -> dipendenti
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $v_1=(1,2,1)$, $v_2=(0,1,1)$, $v_3=(1,3,2)$ sono linearmente indipendenti.

**Ragionamento:**

**Passo 1 — impostiamo l'equazione omogenea**: $\lambda_1(1,2,1)+\lambda_2(0,1,1)+\lambda_3(1,3,2)=(0,0,0)$.

**Passo 2 — trasformiamo in sistema lineare** componente per componente:

$$\begin{cases}\lambda_1+\lambda_3=0\\ 2\lambda_1+\lambda_2+3\lambda_3=0\\ \lambda_1+\lambda_2+2\lambda_3=0\end{cases}$$

**Passo 3 — dalla prima equazione**: $\lambda_1=-\lambda_3$.

**Passo 4 — sostituiamo nella terza equazione**: $-\lambda_3+\lambda_2+2\lambda_3=0$, cioè $\lambda_2=-\lambda_3$.

**Passo 5 — verifichiamo con la seconda equazione** (per controllare la coerenza, esattamente come nell'esercizio del notebook 6.5): $2(-\lambda_3)+(-\lambda_3)+3\lambda_3=-2\lambda_3-\lambda_3+3\lambda_3=0$. Questa equazione è **sempre** soddisfatta, qualunque sia $\lambda_3$: non impone nessuna condizione ulteriore.

**Passo 6 — interpretiamo il risultato.** Abbiamo trovato che, per **ogni** valore di $\lambda_3$, ponendo $\lambda_1=-\lambda_3$ e $\lambda_2=-\lambda_3$ otteniamo una soluzione del sistema omogeneo. Per esempio, con $\lambda_3=1$: $\lambda_1=-1,\lambda_2=-1,\lambda_3=1$, che **non** sono tutti nulli.

**Conclusione:** esiste una combinazione non banale che dà il vettore nullo ($-v_1-v_2+v_3=\mathbf 0$), quindi $v_1,v_2,v_3$ sono **linearmente dipendenti**. Infatti, isolando: $v_3=v_1+v_2$ — verifichiamolo direttamente: $(1,2,1)+(0,1,1)=(1,3,2)=v_3$. ✓ $v_3$ era "ridondante": già ottenibile da $v_1,v_2$.

**Perché il sistema aveva infinite soluzioni invece di una sola:** con 3 incognite e solo 2 equazioni "indipendenti" (la seconda era conseguenza automatica delle altre), il sistema è **sottodeterminato**: resta un parametro libero ($\lambda_3$), segnale sicuro dell'esistenza di soluzioni non banali — e quindi di dipendenza lineare.

````python
import sympy as sp

l1, l2, l3 = sp.symbols('lambda1 lambda2 lambda3')
v1 = sp.Matrix([1, 2, 1])
v2 = sp.Matrix([0, 1, 1])
v3 = sp.Matrix([1, 3, 2])

eq = l1 * v1 + l2 * v2 + l3 * v3
soluzione = sp.solve([eq[0], eq[1], eq[2]], [l1, l2, l3])
print("Soluzione generale del sistema omogeneo:", soluzione)

# Verifica diretta: v3 = v1 + v2 ?
print("v1 + v2 =", (v1 + v2).T, " | v3 =", v3.T)
print("v3 == v1+v2 ?", v1 + v2 == v3)
````

## Riepilogo finale

- $v_1,\dots,v_k$ sono **linearmente indipendenti** se l'unica combinazione lineare che dà $\mathbf 0$ è quella con tutti i coefficienti nulli; altrimenti sono **dipendenti**.
- Dipendenza lineare $\iff$ almeno un vettore è combinazione lineare degli altri: è la formalizzazione precisa dell'idea di "ridondanza" incontrata nel notebook 6.5.
- In pratica, si verifica risolvendo il **sistema lineare omogeneo** $\lambda_1v_1+\cdots+\lambda_kv_k=\mathbf 0$: soluzione unica (nulla) $\Rightarrow$ indipendenti; infinite soluzioni $\Rightarrow$ dipendenti.
- Un insieme che contiene il vettore nullo è sempre dipendente; ogni sottoinsieme di un insieme indipendente è a sua volta indipendente.
