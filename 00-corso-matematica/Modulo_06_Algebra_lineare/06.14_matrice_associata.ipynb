# 6.14 — Matrice associata

**Prerequisiti:** 6.13 Applicazioni lineari, 6.8 Basi.


## Perché ci interessa

Nel notebook 6.13 abbiamo visto che ogni prodotto matrice-vettore $x\mapsto Ax$ è un'applicazione lineare. Ora dimostriamo il viceversa, molto più potente: **ogni** applicazione lineare tra spazi vettoriali di dimensione finita si può rappresentare con una matrice. Questo significa che possiamo studiare le applicazioni lineari — oggetti astratti — usando semplicemente le tecniche di calcolo matriciale già sviluppate (notebook 6.10-6.12): moltiplicazioni, determinanti, sistemi lineari. È il ponte che rende l'algebra lineare astratta effettivamente calcolabile.

## Teoria, passo dopo passo

### L'idea chiave: basta sapere dove vanno i vettori di base

Sia $T:V\to W$ lineare, e sia $\mathcal B=\{v_1,\dots,v_n\}$ una base di $V$ (notebook 6.8). Per ogni $u\in V$, esistono **coordinate uniche** $\lambda_1,\dots,\lambda_n$ tali che $u=\lambda_1v_1+\cdots+\lambda_nv_n$. Applicando $T$ e usando la linearità:

$$T(u) = T(\lambda_1v_1+\cdots+\lambda_nv_n) = \lambda_1T(v_1)+\lambda_2T(v_2)+\cdots+\lambda_nT(v_n)$$

Questo è il fatto cruciale: **$T(u)$ è completamente determinato da $\lambda_1,\dots,\lambda_n$ (le coordinate di $u$) e da $T(v_1),\dots,T(v_n)$ (le immagini dei vettori di base)**. Non serve sapere come $T$ agisce su ogni vettore separatamente: basta conoscerne l'azione sugli $n$ vettori della base, e tutto il resto segue per linearità.

### Costruzione della matrice associata

Fissiamo una base $\mathcal B=\{v_1,\dots,v_n\}$ di $V$ e una base $\mathcal C=\{w_1,\dots,w_m\}$ di $W$. Ogni $T(v_j)$ è un vettore di $W$, quindi ha coordinate uniche rispetto a $\mathcal C$: scriviamo $T(v_j)=a_{1j}w_1+a_{2j}w_2+\cdots+a_{mj}w_m$. La **matrice associata** a $T$ (rispetto a $\mathcal B,\mathcal C$) è la matrice $A$ ($m\times n$) le cui colonne sono proprio queste coordinate: la colonna $j$-esima di $A$ è il vettore delle coordinate di $T(v_j)$ rispetto a $\mathcal C$.

**Perché questa costruzione funziona:** se $x=(\lambda_1,\dots,\lambda_n)$ sono le coordinate di $u$ rispetto a $\mathcal B$, allora — usando il fatto che $Ax$ è la combinazione lineare delle colonne di $A$ con coefficienti $x$ (notebook 6.10) — $Ax$ dà proprio le coordinate di $T(u)=\lambda_1T(v_1)+\cdots+\lambda_nT(v_n)$ rispetto a $\mathcal C$. In sintesi: **moltiplicare per $A$ equivale ad applicare $T$**, lavorando con le coordinate anziché con i vettori astratti.

### Esempio: la matrice della rotazione

Consideriamo $T:\mathbb R^2\to\mathbb R^2$, la rotazione di angolo $\theta$ (notebook 6.13), rispetto alla base canonica $\{e_1,e_2\}$ sia in partenza sia in arrivo. Dove va $e_1=(1,0)$? Ruotandolo di $\theta$: $T(e_1)=(\cos\theta,\sin\theta)$. Dove va $e_2=(0,1)$? $T(e_2)=(-\sin\theta,\cos\theta)$ (ruotato di $\theta$ a partire da "in alto"). La matrice associata ha queste immagini come colonne:

$$A = \begin{pmatrix}\cos\theta & -\sin\theta\\ \sin\theta & \cos\theta\end{pmatrix}$$

esattamente la formula della rotazione anticipata nel notebook 6.13.

### Composizione di applicazioni lineari = prodotto di matrici

Se $T:V\to W$ ha matrice $A$ (rispetto a basi fissate) e $S:W\to U$ ha matrice $B$, la composizione $S\circ T:V\to U$ (notebook 3.7) ha matrice $BA$ (nell'ordine: prima si applica $A$, poi $B$). Questo è esattamente il motivo per cui il prodotto tra matrici è definito nel modo "complicato" visto nel notebook 6.10 (riga per colonna): è costruito apposta per corrispondere alla composizione di funzioni.

````python
import sympy as sp

theta = sp.symbols('theta', real=True)

# Matrice associata alla rotazione: immagini di e1, e2
e1, e2 = sp.Matrix([1, 0]), sp.Matrix([0, 1])

def ruota(v, angolo):
    x, y = v
    return sp.Matrix([x * sp.cos(angolo) - y * sp.sin(angolo),
                       x * sp.sin(angolo) + y * sp.cos(angolo)])

T_e1 = ruota(e1, theta)
T_e2 = ruota(e2, theta)
print("T(e1) =", T_e1.T)
print("T(e2) =", T_e2.T)

A = sp.Matrix.hstack(T_e1, T_e2)
print("\nMatrice associata A =\n", A)

# Verifica: A applicata a un vettore qualsiasi coincide con la rotazione diretta
v = sp.Matrix([3, 1])
print("\nA*v =", (A * v).T)
print("ruota(v) =", ruota(v, theta).T)
print("Coincidono?", sp.simplify(A * v - ruota(v, theta)) == sp.zeros(2, 1))
````

## Esercizio svolto e commentato

**Problema:** trova la matrice associata (rispetto alla base canonica) a $T:\mathbb R^2\to\mathbb R^2$, la riflessione rispetto alla retta $y=x$.

**Ragionamento:**

**Passo 1 — capiamo geometricamente cosa fa la riflessione**: riflettere rispetto alla retta $y=x$ scambia le coordinate: il punto $(a,b)$ va in $(b,a)$.

**Passo 2 — troviamo l'immagine di $e_1=(1,0)$** applicando questa regola: $T(e_1)=T(1,0)=(0,1)$.

**Passo 3 — troviamo l'immagine di $e_2=(0,1)$**: $T(e_2)=T(0,1)=(1,0)$.

**Passo 4 — costruiamo la matrice associata**, usando $T(e_1)$ e $T(e_2)$ come colonne (rispetto alla base canonica, le coordinate di un vettore coincidono con le sue componenti, quindi non serve nessun calcolo aggiuntivo per "tradurre" in coordinate):

$$A = \begin{pmatrix}0&1\\1&0\end{pmatrix}$$

**Verifica:** applichiamo $A$ a un vettore generico $(a,b)$: $A(a,b)=(0\times a+1\times b,\ 1\times a+0\times b)=(b,a)$. ✓ coincide con la regola "scambia le coordinate".

**Perché, lavorando con la base canonica, costruire la matrice associata è così immediato:** le coordinate di un vettore $(x,y)\in\mathbb R^2$ rispetto alla base canonica sono semplicemente $(x,y)$ stesso (nessuna conversione necessaria, a differenza di quanto visto nell'esercizio del notebook 6.8 con basi diverse dalla canonica). Per questo, quando si lavora sempre con basi canoniche, "trovare la matrice associata a $T$" si riduce semplicemente a calcolare $T(e_1),\dots,T(e_n)$ e affiancarli come colonne — senza dover risolvere alcun sistema per trovare le coordinate.

````python
import sympy as sp

a, b = sp.symbols('a b')

A = sp.Matrix([[0, 1], [1, 0]])
v = sp.Matrix([a, b])

print("A*v =", (A * v).T)
print("Regola attesa (scambio coordinate): (b, a)")
````

## Riepilogo finale

- Ogni applicazione lineare $T:V\to W$, tra spazi di dimensione finita, corrisponde a un'unica **matrice associata** $A$, una volta fissate le basi di $V$ e $W$: le colonne di $A$ sono le coordinate delle immagini dei vettori di base.
- Moltiplicare per $A$ equivale ad applicare $T$, lavorando in coordinate anziché con i vettori astratti.
- Con la base canonica, costruire $A$ è immediato: basta calcolare $T(e_1),\dots,T(e_n)$ e affiancarli come colonne.
- La **composizione** di applicazioni lineari corrisponde al **prodotto** delle rispettive matrici (nell'ordine giusto): è il motivo profondo della definizione del prodotto tra matrici.
