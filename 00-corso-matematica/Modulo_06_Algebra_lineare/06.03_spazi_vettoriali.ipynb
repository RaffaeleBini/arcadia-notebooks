# 6.3 — Spazi vettoriali

**Prerequisiti:** 6.1 Vettori, 6.2 Geometria vettoriale, 0.5 Proprietà delle operazioni.


## Perché ci interessa

Finora abbiamo lavorato con vettori come frecce in $\mathbb R^2$: coppie di numeri che si sommano e si moltiplicano per uno scalare. Ma questo "comportamento" — sommare, scalare, con certe regole — non è esclusivo delle frecce. Anche i **polinomi** si sommano e si moltiplicano per un numero (notebook 1.1); anche le **matrici** (che vedremo tra poco); persino le **funzioni** si sommano punto per punto. Se tutte queste strutture diverse condividono le stesse regole di base, conviene astrarle in un unico concetto: lo **spazio vettoriale**. Una volta dimostrato un fatto per uno spazio vettoriale generico, quel fatto vale automaticamente per $\mathbb R^n$, per i polinomi, per le matrici — senza doverlo ridimostrare ogni volta. È il potere dell'astrazione matematica.

## Teoria, passo dopo passo

### Cosa hanno in comune $\mathbb R^2$ e i polinomi?

Riprendiamo la somma di vettori in $\mathbb R^2$ (notebook 6.1) e la somma di polinomi (notebook 1.1). In entrambi i casi valgono le stesse proprietà elementari (notebook 0.5): la somma è **commutativa** ($v+w=w+v$), **associativa** ($(v+w)+u=v+(w+u)$), esiste un **elemento neutro** (il vettore nullo $\mathbf 0$, o il polinomio nullo), e ogni elemento ha un **opposto**. Lo stesso vale per la moltiplicazione per uno scalare: è **distributiva** rispetto alla somma di vettori e rispetto alla somma di scalari.

Questa coincidenza non è un caso: è il segnale che $\mathbb R^2$ e l'insieme dei polinomi sono esempi della **stessa struttura astratta**.

### Definizione di spazio vettoriale

Uno **spazio vettoriale** (su $\mathbb R$) è un insieme $V$, i cui elementi chiamiamo **vettori**, dotato di due operazioni — somma $v+w$ e prodotto per uno scalare $\lambda v$ (con $\lambda\in\mathbb R$) — che soddisfano questi assiomi, per ogni $u,v,w\in V$ e $\lambda,\mu\in\mathbb R$:

1. $(v+w)+u = v+(w+u)$ (associativa della somma)
2. $v+w=w+v$ (commutativa della somma)
3. esiste $\mathbf 0\in V$ tale che $v+\mathbf 0=v$ per ogni $v$ (elemento neutro)
4. per ogni $v$ esiste $-v$ tale che $v+(-v)=\mathbf 0$ (opposto)
5. $\lambda(v+w)=\lambda v+\lambda w$ (distributiva rispetto alla somma di vettori)
6. $(\lambda+\mu)v=\lambda v+\mu v$ (distributiva rispetto alla somma di scalari)
7. $\lambda(\mu v)=(\lambda\mu)v$ (associativa mista)
8. $1\cdot v=v$ (neutralità dello scalare $1$)

Questi otto assiomi sono esattamente la formalizzazione delle proprietà elementari già viste nel notebook 0.5, applicate non più solo ai numeri ma a un insieme $V$ qualunque.

### Esempi di spazi vettoriali

- $\mathbb R^n$ con somma e prodotto per scalare componente per componente (i vettori del notebook 6.1): è l'esempio "prototipo".
- $\mathbb R^n[x]$, l'insieme dei polinomi di grado al più $n$ (notebook 1.1), con la somma usuale di polinomi e il prodotto per uno scalare: è uno spazio vettoriale (i coefficienti si sommano e si scalano come se fossero le componenti di un vettore).
- L'insieme delle matrici $m\times n$ (che studieremo nel notebook 6.10), con somma elemento per elemento.
- L'insieme delle funzioni $f:\mathbb R\to\mathbb R$, con $(f+g)(x):=f(x)+g(x)$ e $(\lambda f)(x):=\lambda f(x)$: anche questo, sorprendentemente, è uno spazio vettoriale — infinito-dimensionale, come vedremo nel notebook 6.9.

### Un esempio di "non" spazio vettoriale

Non tutti gli insiemi con operazioni "simili" formano uno spazio vettoriale. Consideriamo $V=\{(x,y): x\ge0,\,y\ge0\}$ (il primo quadrante del piano), con la somma e il prodotto per scalare usuali. La somma di due elementi di $V$ resta in $V$ (se $x_1,x_2\ge0$ allora $x_1+x_2\ge0$), ma l'assioma 4 (esistenza dell'opposto) fallisce: preso $v=(1,1)\in V$, il suo opposto $-v=(-1,-1)$ **non** appartiene a $V$ (ha componenti negative). Quindi $V$, con queste operazioni, **non** è uno spazio vettoriale.

````python
import sympy as sp

# Verifica simbolica di alcuni assiomi per lo spazio dei polinomi di grado <= 2
x, l, m = sp.symbols('x lambda mu')
a0, a1, a2, b0, b1, b2, c0, c1, c2 = sp.symbols('a0 a1 a2 b0 b1 b2 c0 c1 c2')

p = a0 + a1 * x + a2 * x**2
q = b0 + b1 * x + b2 * x**2
r = c0 + c1 * x + c2 * x**2

# Assioma 2: commutativa della somma
print("p+q == q+p ?", sp.expand(p + q - (q + p)) == 0)

# Assioma 5: distributiva rispetto alla somma di vettori (polinomi)
lhs5 = sp.expand(l * (p + q))
rhs5 = sp.expand(l * p + l * q)
print("lambda(p+q) == lambda*p + lambda*q ?", lhs5 == rhs5)

# Assioma 6: distributiva rispetto alla somma di scalari
lhs6 = sp.expand((l + m) * p)
rhs6 = sp.expand(l * p + m * p)
print("(lambda+mu)*p == lambda*p + mu*p ?", lhs6 == rhs6)

# Assioma 7: associativa mista
lhs7 = sp.expand(l * (m * p))
rhs7 = sp.expand((l * m) * p)
print("lambda*(mu*p) == (lambda*mu)*p ?", lhs7 == rhs7)
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $V=\{(x,y,z)\in\mathbb R^3 : x+y+z=0\}$, con somma e prodotto per scalare usuali (come in $\mathbb R^3$), è uno spazio vettoriale.

**Ragionamento:**

**Passo 1 — invece di verificare tutti e 8 gli assiomi da zero, sfruttiamo il fatto che $V\subset\mathbb R^3$** e che $\mathbb R^3$, con le stesse operazioni, è già uno spazio vettoriale (esempio visto sopra). Gli assiomi 1, 2, 5, 6, 7, 8 (quelli che riguardano solo il "comportamento" delle operazioni, non l'appartenenza a $V$) valgono automaticamente in $V$ perché valgono già in tutto $\mathbb R^3$: se un'identità come $v+w=w+v$ è vera per **tutti** i vettori di $\mathbb R^3$, resta vera in particolare per quelli di $V$.

**Passo 2 — resta da controllare solo se $V$ è "chiuso" rispetto alle operazioni** (cioè se sommando o scalando elementi di $V$ si resta dentro $V$) e se contiene l'elemento neutro e gli opposti. Questo anticipa il criterio generale che formalizzeremo nel prossimo notebook (6.4, sottospazi): per ora verifichiamolo a mano.

**Passo 3 — chiusura rispetto alla somma.** Siano $v=(x_1,y_1,z_1)$ e $w=(x_2,y_2,z_2)$ in $V$, cioè $x_1+y_1+z_1=0$ e $x_2+y_2+z_2=0$. La loro somma è $v+w=(x_1+x_2,\,y_1+y_2,\,z_1+z_2)$; sommando le due componenti otteniamo $(x_1+x_2)+(y_1+y_2)+(z_1+z_2)=(x_1+y_1+z_1)+(x_2+y_2+z_2)=0+0=0$. Quindi $v+w\in V$: la somma è chiusa.

**Passo 4 — chiusura rispetto al prodotto per scalare.** Se $x+y+z=0$, allora $\lambda x+\lambda y+\lambda z=\lambda(x+y+z)=\lambda\cdot0=0$ (raccogliendo $\lambda$ a fattor comune, distributiva, notebook 0.5): quindi $\lambda v\in V$.

**Passo 5 — elemento neutro e opposti.** $\mathbf 0=(0,0,0)$ soddisfa $0+0+0=0$, quindi $\mathbf 0\in V$. Se $v=(x,y,z)\in V$, allora $-v=(-x,-y,-z)$ soddisfa $-x-y-z=-(x+y+z)=0$, quindi $-v\in V$.

**Conclusione:** $V$ è uno spazio vettoriale (è un sottospazio di $\mathbb R^3$, come chiariremo nel notebook 6.4).

````python
import sympy as sp

x1, y1, z1, x2, y2, z2, l = sp.symbols('x1 y1 z1 x2 y2 z2 lambda')

# Imponiamo che v=(x1,y1,z1) e w=(x2,y2,z2) soddisfino x+y+z=0
# (sostituendo z1 = -x1-y1, z2 = -x2-y2)
z1_expr = -x1 - y1
z2_expr = -x2 - y2

somma_componenti = (x1 + x2) + (y1 + y2) + (z1_expr + z2_expr)
print("Somma delle componenti di v+w:", sp.simplify(somma_componenti))  # deve fare 0

scalato = l * x1 + l * y1 + l * z1_expr
print("Somma delle componenti di lambda*v:", sp.simplify(scalato))  # deve fare 0
````

## Riepilogo finale

- Uno **spazio vettoriale** è un insieme dotato di somma e prodotto per scalare che soddisfano 8 assiomi — la generalizzazione astratta delle proprietà già viste per i numeri (notebook 0.5) e per i vettori di $\mathbb R^2$ (notebook 6.1).
- $\mathbb R^n$, i polinomi, le matrici e le funzioni sono tutti esempi di spazi vettoriali: strutture apparentemente diverse che condividono le stesse regole algebriche.
- Per verificare se un sottoinsieme di uno spazio vettoriale già noto è a sua volta uno spazio vettoriale, basta controllare la **chiusura** rispetto alle due operazioni e la presenza dell'elemento neutro e degli opposti — non serve riverificare tutti gli 8 assiomi da zero.
- Non ogni insieme "simile" a un vettore è uno spazio vettoriale: basta che manchi un solo assioma (come l'opposto, per $\mathbb N$, o la chiusura, per una retta non passante per l'origine).
