# 6.24 — Teorema spettrale

**Prerequisiti:** 6.23 Forme quadratiche, 6.22 Basi ortonormali, 6.19 Diagonalizzazione, 6.17 Autovalori.


## Perché ci interessa

Nel notebook 6.19 abbiamo visto che una matrice è diagonalizzabile solo a certe condizioni (autovalori distinti, o autospazi di dimensione sufficiente) — non tutte le matrici lo sono, come l'esempio del notebook 6.19 ($\begin{pmatrix}0&1\\0&0\end{pmatrix}$) mostrava. Nel notebook 6.23 abbiamo però **anticipato** un risultato notevole: le matrici **simmetriche** sono sempre diagonalizzabili, e in un modo particolarmente elegante — con una matrice ortogonale. Questo è il **teorema spettrale**, uno dei risultati più belli e utili dell'algebra lineare elementare, che chiude il cerchio tra autovalori, ortogonalità e forme quadratiche, i tre temi conclusivi di questo modulo.

## Teoria, passo dopo passo

### Enunciato del teorema

> **Teorema spettrale (caso reale, simmetrico).** Se $A$ è una matrice simmetrica reale ($A=A^T$), allora:
> 1. tutti gli autovalori di $A$ sono **reali**;
> 2. autovettori relativi ad autovalori **distinti** sono **ortogonali** (non solo indipendenti, come nel caso generale del notebook 6.18);
> 3. $A$ è **sempre diagonalizzabile**, ed esiste una matrice **ortogonale** $Q$ (notebook 6.22) tale che $A=QDQ^T$, con $D$ diagonale contenente gli autovalori (reali) di $A$.

### Dimostrazione del punto 2 (autovettori ortogonali)

Siano $v_1,v_2$ autovettori di $A$ con autovalori distinti $\lambda_1\ne\lambda_2$: $Av_1=\lambda_1v_1$, $Av_2=\lambda_2v_2$. Consideriamo il prodotto scalare $\langle Av_1,v_2\rangle$, e calcoliamolo in due modi.

**Primo modo**: $\langle Av_1,v_2\rangle=\langle\lambda_1v_1,v_2\rangle=\lambda_1\langle v_1,v_2\rangle$ (bilinearità, notebook 6.20).

**Secondo modo**: usando che, per matrici simmetriche, $\langle Av_1,v_2\rangle=(Av_1)^Tv_2=v_1^TA^Tv_2=v_1^TAv_2=\langle v_1,Av_2\rangle$ (usando $A^T=A$) $=\langle v_1,\lambda_2v_2\rangle=\lambda_2\langle v_1,v_2\rangle$.

**Confrontando i due modi**: $\lambda_1\langle v_1,v_2\rangle=\lambda_2\langle v_1,v_2\rangle$, cioè $(\lambda_1-\lambda_2)\langle v_1,v_2\rangle=0$. Poiché $\lambda_1\ne\lambda_2$ (ipotesi), deve essere $\langle v_1,v_2\rangle=0$: **$v_1$ e $v_2$ sono ortogonali**. $\blacksquare$

Questo risultato è il cuore del teorema: nel caso generale (notebook 6.18), autovalori distinti garantivano solo l'**indipendenza**; per matrici simmetriche, garantiscono l'**ortogonalità**, una proprietà molto più forte.

### Perché $A=QDQ^T$ (conseguenza pratica)

Per costruire $Q$: si trovano gli autovettori di $A$ per ciascun autovalore; per il punto 2, autovettori relativi ad autovalori distinti sono già ortogonali; se un autovalore ha molteplicità $>1$, si applica Gram-Schmidt (notebook 6.21) **all'interno** del suo autospazio, per ottenere anche lì vettori ortogonali tra loro (restando comunque autovettori per lo stesso autovalore). Si normalizzano tutti gli autovettori (dividendo per la norma) per ottenere una base **ortonormale** di autovettori $q_1,\dots,q_n$, e si affiancano come colonne di $Q$ (che risulta ortogonale, per costruzione, notebook 6.22). La formula $A=PDP^{-1}$ del notebook 6.19, con $P=Q$ ortogonale, diventa $A=QDQ^{-1}=QDQ^T$ (usando $Q^{-1}=Q^T$, notebook 6.22).

### Il cerchio si chiude: forme quadratiche

Questo è esattamente il risultato usato (anticipato) nel notebook 6.23: ogni forma quadratica $q(v)=v^TAv$, con $A$ simmetrica, si riduce — cambiando base con la matrice ortogonale $Q$ del teorema spettrale — a una somma di quadrati pesati dagli autovalori (reali, garantiti dal punto 1) di $A$: $q(Qw)=\lambda_1w_1^2+\cdots+\lambda_nw_n^2$. La classificazione delle forme quadratiche del notebook precedente **è** un'applicazione diretta del teorema spettrale.

````python
import sympy as sp

A = sp.Matrix([[4, 2], [2, 1]])  # simmetrica
print("A è simmetrica?", A == A.T)

# Autovalori (garantiti reali dal teorema spettrale)
print("Autovalori:", A.eigenvals())

# Diagonalizzazione ortogonale con SymPy
Q, D = A.diagonalize(normalize=True)
print("\nQ (ortogonale, colonne = autovettori normalizzati) =\n", Q)
print("\nD =\n", D)

# Verifica Q ortogonale: Q^T Q = I
print("\nQ^T Q =\n", sp.simplify(Q.T * Q))

# Verifica A = Q D Q^T
ricostruita = sp.simplify(Q * D * Q.T)
print("\nQ D Q^T =\n", ricostruita)
print("Coincide con A?", ricostruita == A)
````

## Esercizio svolto e commentato

**Problema:** applica il teorema spettrale a $A=\begin{pmatrix}3&1\\1&3\end{pmatrix}$ (simmetrica): trova $Q$ e $D$, e verifica $A=QDQ^T$.

**Ragionamento:**

**Passo 1 — verifichiamo che $A$ è simmetrica**: $a_{12}=1=a_{21}$. ✓ Il teorema spettrale si applica.

**Passo 2 — troviamo gli autovalori**: $\operatorname{tr}(A)=6$, $\det A=9-1=8$. $\lambda^2-6\lambda+8=0$: $\lambda=\frac{6\pm\sqrt{36-32}}2=\frac{6\pm2}2$, cioè $\lambda_1=4,\lambda_2=2$ (entrambi reali, come garantito dal punto 1 del teorema).

**Passo 3 — troviamo gli autovettori**. Per $\lambda_1=4$: $A-4I=\begin{pmatrix}-1&1\\1&-1\end{pmatrix}$, sistema $-x+y=0\Rightarrow x=y$: autovettore $(1,1)$. Per $\lambda_2=2$: $A-2I=\begin{pmatrix}1&1\\1&1\end{pmatrix}$, sistema $x+y=0\Rightarrow x=-y$: autovettore $(1,-1)$.

**Passo 4 — verifichiamo l'ortogonalità** (garantita dal punto 2 del teorema, essendo $\lambda_1\ne\lambda_2$): $(1,1)\cdot(1,-1)=1-1=0$. ✓ Confermato senza sorprese.

**Passo 5 — normalizziamo** per ottenere una base ortonormale: $\|(1,1)\|=\sqrt2$, quindi $q_1=\left(\frac1{\sqrt2},\frac1{\sqrt2}\right)$; $\|(1,-1)\|=\sqrt2$, quindi $q_2=\left(\frac1{\sqrt2},-\frac1{\sqrt2}\right)$.

**Passo 6 — componiamo**: $Q=\begin{pmatrix}\frac1{\sqrt2}&\frac1{\sqrt2}\\\frac1{\sqrt2}&-\frac1{\sqrt2}\end{pmatrix}$, $D=\begin{pmatrix}4&0\\0&2\end{pmatrix}$.

**Conclusione:** $A=QDQ^T$, verificabile esplicitamente moltiplicando le tre matrici.

**Perché il fatto che gli autovettori fossero *già* ortogonali (passo 4), senza bisogno di applicare Gram-Schmidt, non è un caso fortunato ma una conseguenza diretta della simmetria di $A$:** avendo trovato autovalori **distinti** ($4\ne2$), il punto 2 del teorema spettrale **garantisce** l'ortogonalità dei rispettivi autovettori — non serve verificarla "per fortuna", è una conseguenza matematica certa. L'ortogonalizzazione di Gram-Schmidt sarebbe stata necessaria solo nel caso di un autovalore con molteplicità maggiore di 1 (autospazio di dimensione $>1$), per ottenere una base ortogonale **all'interno** di quell'autospazio.

````python
import sympy as sp

A = sp.Matrix([[3, 1], [1, 3]])
print("A simmetrica?", A == A.T)
print("Autovalori:", A.eigenvals())

Q, D = A.diagonalize(normalize=True)
print("\nQ =\n", Q)
print("\nD =\n", D)
print("\nQ^T Q =\n", sp.simplify(Q.T * Q))
print("\nQ D Q^T =\n", sp.simplify(Q * D * Q.T))
print("Coincide con A?", sp.simplify(Q * D * Q.T - A) == sp.zeros(2, 2))
````

## Riepilogo finale

- Il **teorema spettrale**: ogni matrice simmetrica reale ha autovalori reali, autovettori relativi ad autovalori distinti ortogonali, ed è diagonalizzabile con una matrice **ortogonale**: $A=QDQ^T$.
- Questo risultato spiega perché, nel notebook 6.23, le forme quadratiche (associate a matrici simmetriche) si riducono sempre a somme di quadrati senza termini misti, tramite un cambio di coordinate che non deforma lunghezze o angoli.
- Il teorema spettrale è il risultato che chiude il cerchio tra autovalori (notebook 6.17-6.19), ortogonalità (notebook 6.20-6.22) e forme quadratiche (notebook 6.23): tre fili dello stesso modulo che si intrecciano in un unico teorema.
