## Perché ci interessa

Nel notebook 2.1 abbiamo imparato a descrivere un **punto** del piano con una coppia di numeri $(x,y)$. Ma molte grandezze della fisica e della matematica non sono punti: sono **spostamenti**, **velocità**, **forze** — cose che hanno sia una lunghezza sia una direzione. Se dico "sono a 10 km da qui" non ho detto tutto: manca la direzione. Se dico "10 km verso nord-est" ho dato un'informazione completa.

Questa idea — una quantità con lunghezza e direzione — si chiama **vettore**. È lo strumento che ci permette di sommare spostamenti, scalare forze, e più in generale di fare "aritmetica" con oggetti più ricchi dei singoli numeri. È il mattone fondamentale di tutta l'algebra lineare, che occuperà questo intero modulo.

## Teoria, passo dopo passo

### Dal punto al vettore: uno spostamento

Consideriamo due punti del piano cartesiano (notebook 2.1): $A=(1,1)$ e $B=(4,3)$. Lo **spostamento** per andare da $A$ a $B$ si misura sottraendo le coordinate:

$$\overrightarrow{AB} = (4-1,\ 3-1) = (3,2)$$

Questo spostamento — "3 unità a destra, 2 in su" — è lo stesso indipendentemente da dove parto: se partissi da $A'=(0,0)$ arriverei a $B'=(3,2)$. È proprio questa indipendenza dal punto di partenza a definire un **vettore**: un vettore non è "attaccato" a un punto preciso, rappresenta solo una direzione e una lunghezza.

### Definizione: vettore in $\mathbb R^2$ e $\mathbb R^n$

Un **vettore** di $\mathbb R^2$ è una coppia ordinata di numeri reali $v=(v_1,v_2)$. Più in generale, un vettore di $\mathbb R^n$ è una $n$-upla ordinata $v=(v_1,v_2,\dots,v_n)$, con $v_i\in\mathbb R$ detti **componenti** (o coordinate) del vettore.

Graficamente, in $\mathbb R^2$ o $\mathbb R^3$, disegniamo un vettore come una **freccia**: la si può disegnare partendo da un punto qualsiasi, ma per convenzione disegniamo spesso il **vettore posizione**, con la coda nell'origine $(0,0)$ e la punta nel punto $(v_1,v_2)$.

### Somma di vettori

Dati $v=(v_1,v_2)$ e $w=(w_1,w_2)$, la loro **somma** è definita componente per componente:

$$v+w = (v_1+w_1,\ v_2+w_2)$$

Geometricamente corrisponde alla **regola del parallelogramma**: se disegniamo $v$ e $w$ dalla stessa origine, la somma $v+w$ è la diagonale del parallelogramma che hanno come lati. Equivalentemente (regola "punta-coda"): disegniamo $v$, poi disegniamo $w$ a partire dalla punta di $v$; il vettore che va dall'origine alla punta di $w$ è $v+w$.

**Perché ha senso questa definizione:** se $v=\overrightarrow{AB}$ è uno spostamento e $w=\overrightarrow{BC}$ è lo spostamento successivo, lo spostamento complessivo $\overrightarrow{AC}$ ha componenti $(C_1-A_1,\,C_2-A_2)=(C_1-B_1+B_1-A_1,\,\dots)=(B_1-A_1+C_1-B_1,\dots)$, cioè esattamente la somma componente per componente di $\overrightarrow{AB}$ e $\overrightarrow{BC}$: la definizione algebrica cattura fedelmente l'idea intuitiva di "spostamenti concatenati".

### Moltiplicazione per uno scalare

Dato un numero reale $\lambda$ (detto **scalare**) e un vettore $v=(v_1,v_2)$, il prodotto è:

$$\lambda v = (\lambda v_1,\ \lambda v_2)$$

Geometricamente: $\lambda v$ ha la stessa direzione di $v$ se $\lambda>0$ (stessa freccia, allungata o accorciata di un fattore $|\lambda|$), direzione opposta se $\lambda<0$, e $\lambda v=(0,0)$ se $\lambda=0$. Il caso $\lambda=-1$ dà il vettore **opposto** $-v$, usato per definire la differenza $v-w := v+(-w)$.

### Il vettore nullo

Il vettore $\mathbf 0=(0,0,\dots,0)$ (tutte le componenti nulle) è l'**elemento neutro** della somma (notebook 0.5): $v+\mathbf 0=v$ per ogni $v$. Geometricamente è una freccia di lunghezza zero, "senza direzione".

````python
import numpy as np
import matplotlib.pyplot as plt

# Due vettori in R^2
v = np.array([3, 2])
w = np.array([1, -1])
somma = v + w

print("v =", v)
print("w =", w)
print("v + w =", somma)
print("2*v =", 2 * v)
print("-v =", -v)

# Visualizziamo v, w e v+w con la regola del parallelogramma
fig, ax = plt.subplots(figsize=(6, 6))
origine = np.array([0, 0])

ax.quiver(*origine, *v, angles='xy', scale_units='xy', scale=1, color='blue', label='v')
ax.quiver(*origine, *w, angles='xy', scale_units='xy', scale=1, color='green', label='w')
ax.quiver(*origine, *somma, angles='xy', scale_units='xy', scale=1, color='red', label='v+w')
# lato "punta-coda": w spostato sulla punta di v
ax.quiver(*v, *w, angles='xy', scale_units='xy', scale=1, color='green', alpha=0.4, linestyle='dashed')

ax.set_xlim(-1, 5)
ax.set_ylim(-2, 4)
ax.axhline(0, color='gray', linewidth=0.5)
ax.axvline(0, color='gray', linewidth=0.5)
ax.legend()
ax.set_aspect('equal')
ax.set_title("Somma di vettori: regola del parallelogramma")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dati i punti $A=(2,1)$, $B=(5,5)$, $C=(6,2)$, verifica che $\overrightarrow{AB}+\overrightarrow{BC}=\overrightarrow{AC}$.

**Ragionamento:**

**Passo 1 — calcoliamo $\overrightarrow{AB}$**, sottraendo le coordinate del punto di partenza da quelle di arrivo (come visto sopra): $\overrightarrow{AB}=(5-2,\,5-1)=(3,4)$.

**Passo 2 — calcoliamo $\overrightarrow{BC}$** allo stesso modo: $\overrightarrow{BC}=(6-5,\,2-5)=(1,-3)$.

**Passo 3 — sommiamo i due vettori** componente per componente: $\overrightarrow{AB}+\overrightarrow{BC}=(3+1,\,4-3)=(4,1)$.

**Passo 4 — calcoliamo direttamente $\overrightarrow{AC}$** dai punti estremi: $\overrightarrow{AC}=(6-2,\,2-1)=(4,1)$.

**Conclusione:** i due risultati coincidono, $(4,1)=(4,1)$. Questo conferma concretamente il ragionamento generale fatto sopra: concatenare gli spostamenti $A\to B$ e $B\to C$ equivale allo spostamento diretto $A\to C$, **qualunque** sia il punto intermedio $B$ — è la proprietà che rende gli spostamenti "coerenti" indipendentemente dal percorso scelto per raggiungere la destinazione.

````python
import numpy as np

A = np.array([2, 1])
B = np.array([5, 5])
C = np.array([6, 2])

AB = B - A
BC = C - B
AC = C - A

print("AB =", AB)
print("BC =", BC)
print("AB + BC =", AB + BC)
print("AC =", AC)
print("Coincidono?", np.array_equal(AB + BC, AC))
````

## Riepilogo finale

- Un vettore rappresenta una **direzione e una lunghezza**, non un punto fisso: due frecce parallele, della stessa lunghezza e verso, sono lo stesso vettore ovunque siano disegnate.
- In $\mathbb R^n$, un vettore è una $n$-upla di numeri $v=(v_1,\dots,v_n)$; in $\mathbb R^2$ possiamo visualizzarlo come freccia dall'origine al punto $(v_1,v_2)$.
- La **somma** si fa componente per componente e corrisponde geometricamente alla regola del parallelogramma (o "punta-coda").
- La **moltiplicazione per uno scalare** $\lambda v$ allunga/accorcia (e, se $\lambda<0$, inverte) la freccia.
- Il vettore $\overrightarrow{AB}$ tra due punti si calcola sottraendo le coordinate: $\overrightarrow{AB}=B-A$.
