## Perché ci interessa

Nel notebook 6.4 abbiamo imparato a riconoscere se un insieme dato è un sottospazio. Ora ci poniamo la domanda opposta e, per certi versi, più costruttiva: partendo da alcuni vettori fissati, quali **altri** vettori possiamo ottenere sommandoli e scalandoli? Questa domanda è tutt'altro che astratta: se $v_1,v_2$ sono due "ingredienti" base (due colori primari, due forze applicate, due direzioni di moto), sapere quali vettori si possono "fabbricare" a partire da loro è centrale in fisica, computer graphics, e — come vedremo — nella risoluzione dei sistemi lineari (notebook 1.7), che altro non sono che la domanda "un dato vettore è ottenibile combinando le colonne di una matrice?".

## Teoria, passo dopo passo

### Definizione di combinazione lineare

Dati i vettori $v_1,v_2,\dots,v_k$ di uno spazio vettoriale $V$, una **combinazione lineare** di essi è un vettore della forma:

$$\lambda_1 v_1+\lambda_2 v_2+\cdots+\lambda_k v_k$$

dove $\lambda_1,\dots,\lambda_k\in\mathbb R$ sono detti **coefficienti** della combinazione. Per esempio, in $\mathbb R^2$, il vettore $(7,4)$ è combinazione lineare di $v_1=(1,0)$ e $v_2=(0,1)$ con coefficienti $\lambda_1=7,\lambda_2=4$: infatti $7(1,0)+4(0,1)=(7,4)$.

### Trovare i coefficienti: un sistema lineare

Data una lista di vettori $v_1,\dots,v_k$ e un vettore $u$, chiedersi "$u$ è combinazione lineare di $v_1,\dots,v_k$?" equivale a chiedersi se esistono $\lambda_1,\dots,\lambda_k$ tali che $\lambda_1v_1+\cdots+\lambda_kv_k=u$. Scrivendo questa equazione componente per componente, otteniamo esattamente un **sistema lineare** (notebook 1.7) nelle incognite $\lambda_1,\dots,\lambda_k$: se il sistema ha soluzione, $u$ è combinazione lineare; se non ne ha, non lo è.

**Esempio:** $u=(4,5)$ è combinazione lineare di $v_1=(1,1)$ e $v_2=(1,-1)$? Impostiamo $\lambda_1(1,1)+\lambda_2(1,-1)=(4,5)$, cioè il sistema $\begin{cases}\lambda_1+\lambda_2=4\\\lambda_1-\lambda_2=5\end{cases}$. Sommando le due equazioni: $2\lambda_1=9$, quindi $\lambda_1=4.5$, e di conseguenza $\lambda_2=-0.5$. Il sistema ha soluzione: sì, $u$ è combinazione lineare di $v_1,v_2$.

### Il concetto di span (inviluppo lineare)

L'insieme di **tutte** le possibili combinazioni lineari di $v_1,\dots,v_k$ si chiama **span** (o inviluppo lineare, o sottospazio generato) e si indica $\operatorname{span}(v_1,\dots,v_k)$:

$$\operatorname{span}(v_1,\dots,v_k) = \{\lambda_1v_1+\cdots+\lambda_kv_k : \lambda_1,\dots,\lambda_k\in\mathbb R\}$$

**Fatto chiave: lo span è sempre un sottospazio.** Verifichiamolo con il criterio del notebook 6.4: (1) il vettore nullo si ottiene con tutti i coefficienti nulli, $0\cdot v_1+\cdots+0\cdot v_k=\mathbf 0$, quindi $\mathbf 0\in\operatorname{span}(v_1,\dots,v_k)$. (2) La somma di due combinazioni lineari è ancora una combinazione lineare: $(\lambda_1v_1+\cdots+\lambda_kv_k)+(\mu_1v_1+\cdots+\mu_kv_k)=(\lambda_1+\mu_1)v_1+\cdots+(\lambda_k+\mu_k)v_k$ (raggruppando, distributiva, notebook 0.5). (3) Analogamente, uno scalare per una combinazione lineare è ancora una combinazione lineare: $\alpha(\lambda_1v_1+\cdots+\lambda_kv_k)=(\alpha\lambda_1)v_1+\cdots+(\alpha\lambda_k)v_k$.

### Esempi geometrici

- $\operatorname{span}((1,0))=\{(\lambda,0):\lambda\in\mathbb R\}$: l'intero asse $x$, una retta.
- $\operatorname{span}((1,0),(0,1))=\mathbb R^2$: con due vettori non allineati si può raggiungere **ogni** punto del piano (lo si vede dal sistema lineare sempre risolubile, come nell'esempio sopra).
- $\operatorname{span}((1,1),(2,2))$: poiché $(2,2)=2(1,1)$, il secondo vettore non aggiunge nulla di nuovo; lo span resta la retta $\operatorname{span}((1,1))$. Questa "ridondanza" è esattamente ciò che il prossimo notebook (6.6, indipendenza lineare) formalizzerà.

````python
import sympy as sp

# u = (4,5) è combinazione lineare di v1=(1,1), v2=(1,-1)?
l1, l2 = sp.symbols('lambda1 lambda2')
v1 = sp.Matrix([1, 1])
v2 = sp.Matrix([1, -1])
u = sp.Matrix([4, 5])

eq = l1 * v1 + l2 * v2 - u
soluzione = sp.solve([eq[0], eq[1]], [l1, l2])
print("Coefficienti trovati:", soluzione)
print("Verifica:", soluzione[l1] * v1 + soluzione[l2] * v2)

# span((1,0),(0,1)) copre tutto R^2: verifichiamo con un vettore "casuale"
target = sp.Matrix([-3, 7])
e1 = sp.Matrix([1, 0])
e2 = sp.Matrix([0, 1])
c1, c2 = sp.symbols('c1 c2')
sol2 = sp.solve([(c1*e1 + c2*e2 - target)[0], (c1*e1 + c2*e2 - target)[1]], [c1, c2])
print("\nCoefficienti per (-3,7) con base canonica:", sol2)
````

## Esercizio svolto e commentato

**Problema:** il vettore $u=(3,-1,5)$ è combinazione lineare di $v_1=(1,0,2)$, $v_2=(0,1,1)$?

**Ragionamento:**

**Passo 1 — impostiamo l'equazione vettoriale**: cerchiamo $\lambda_1,\lambda_2$ tali che $\lambda_1(1,0,2)+\lambda_2(0,1,1)=(3,-1,5)$.

**Passo 2 — trasformiamola in un sistema lineare**, uguagliando componente per componente (notebook 1.7):

$$\begin{cases}\lambda_1 = 3\\ \lambda_2 = -1\\ 2\lambda_1+\lambda_2 = 5\end{cases}$$

**Passo 3 — le prime due equazioni ci danno già $\lambda_1,\lambda_2$ direttamente** (un vantaggio della struttura semplice di $v_1,v_2$, che hanno "1" in posizioni diverse): $\lambda_1=3$, $\lambda_2=-1$.

**Passo 4 — verifichiamo con la terza equazione**, che è la condizione "extra" da controllare (il sistema ha 3 equazioni ma solo 2 incognite: è un sistema **sovradeterminato**, che potrebbe non avere soluzione): $2(3)+(-1)=6-1=5$. ✓ coincide con il membro destro.

**Conclusione:** sì, $u=3v_1-v_2$ è combinazione lineare di $v_1,v_2$; se la terza equazione **non** fosse stata soddisfatta, avremmo concluso che $u\notin\operatorname{span}(v_1,v_2)$ (il sistema sarebbe stato incompatibile).

**Perché il caso "più equazioni che incognite" è tipico qui:** lo span di due vettori in $\mathbb R^3$ è, in generale, solo un **piano** (un sottoinsieme "piccolo" di $\mathbb R^3$, di dimensione 2 in uno spazio di dimensione 3, concetti che formalizzeremo nel notebook 6.9): non tutti i vettori di $\mathbb R^3$ vi appartengono, ed è normale che il sistema, con 3 equazioni per solo 2 incognite, sia "sovravincolato" e possa risultare incompatibile per un $u$ generico.

````python
import sympy as sp

l1, l2 = sp.symbols('lambda1 lambda2')
v1 = sp.Matrix([1, 0, 2])
v2 = sp.Matrix([0, 1, 1])
u = sp.Matrix([3, -1, 5])

eq = l1 * v1 + l2 * v2 - u
soluzione = sp.solve([eq[0], eq[1], eq[2]], [l1, l2])
print("Soluzione:", soluzione)
if soluzione:
    print("u è combinazione lineare di v1, v2 con coefficienti:", soluzione)
else:
    print("u NON è combinazione lineare di v1, v2 (sistema incompatibile)")
````

## Riepilogo finale

- Una **combinazione lineare** di $v_1,\dots,v_k$ è un vettore $\lambda_1v_1+\cdots+\lambda_kv_k$; trovare i coefficienti equivale a risolvere un sistema lineare (notebook 1.7).
- Lo **span** $\operatorname{span}(v_1,\dots,v_k)$, l'insieme di tutte le combinazioni lineari possibili, è sempre un sottospazio — anzi, è il più piccolo sottospazio che contiene tutti i $v_i$.
- Geometricamente, lo span di uno o più vettori in $\mathbb R^2$/$\mathbb R^3$ dà rette, piani, o l'intero spazio, a seconda di quanti vettori "indipendenti" ci sono tra quelli scelti.
- Alcuni vettori possono essere "ridondanti" nello span (come $(2,2)$ rispetto a $(1,1)$): il prossimo notebook formalizza quando questo accade.
