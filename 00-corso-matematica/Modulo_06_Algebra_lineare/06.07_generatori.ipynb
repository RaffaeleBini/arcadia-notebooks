# 6.7 — Generatori

**Prerequisiti:** 6.5 Combinazioni lineari, 6.6 Indipendenza lineare.


## Perché ci interessa

Nel notebook 6.5 abbiamo visto che $\operatorname{span}((1,0),(0,1))=\mathbb R^2$: con questi due vettori possiamo "raggiungere" ogni punto del piano. Diciamo allora che $(1,0),(0,1)$ **generano** $\mathbb R^2$. Ma non è l'unica coppia che funziona: anche $(1,1),(1,-1)$ generano $\mathbb R^2$ (lo abbiamo verificato nell'esempio del notebook 6.5). Ci chiediamo ora: cosa serve, in generale, perché un insieme di vettori "copra" un intero spazio vettoriale? E qual è il numero minimo di vettori necessario? Rispondere a queste domande ci porterà dritti al concetto di **base**, nel prossimo notebook.

## Teoria, passo dopo passo

### Definizione: insieme di generatori

Un insieme di vettori $\{v_1,\dots,v_k\}$ di uno spazio vettoriale $V$ si dice **insieme di generatori** di $V$ (o che $v_1,\dots,v_k$ **generano** $V$) se ogni vettore di $V$ è combinazione lineare di $v_1,\dots,v_k$, cioè se:

$$\operatorname{span}(v_1,\dots,v_k)=V$$

Uno spazio vettoriale che ammette un insieme finito di generatori si dice **finitamente generato**: in questo modulo lavoreremo quasi sempre in questo caso (come $\mathbb R^n$ o i polinomi di grado limitato); un'eccezione notevole è lo spazio di **tutte** le funzioni, che non è finitamente generato.

### Come verificare che un insieme genera $V$

Per $V=\mathbb R^n$, verificare che $v_1,\dots,v_k$ generano $\mathbb R^n$ equivale a verificare che, per **ogni** vettore $u\in\mathbb R^n$ (arbitrario, con componenti generiche), il sistema lineare $\lambda_1v_1+\cdots+\lambda_kv_k=u$ ammette sempre soluzione — non solo per uno specifico $u$, come facevamo nel notebook 6.5, ma per un $u$ "generico".

### Generatori "in eccesso": si può sempre ridurre

Se $v_1,\dots,v_k$ generano $V$ ma sono linearmente **dipendenti** (notebook 6.6), allora almeno uno di essi — diciamo $v_j$ — è combinazione lineare degli altri. In questo caso possiamo **eliminare $v_j$** dalla lista senza perdere nulla: ogni combinazione lineare che coinvolgeva $v_j$ può essere riscritta sostituendo $v_j$ con la sua espressione in funzione degli altri, ottenendo comunque una combinazione lineare dei rimanenti $v_1,\dots,\widehat{v_j},\dots,v_k$ (il cappuccio indica "vettore omesso"). Quindi $\operatorname{span}(v_1,\dots,\widehat{v_j},\dots,v_k)$ resta uguale a $V$: **i generatori restanti generano ancora tutto $V$**.

**Esempio:** $(1,0),(0,1),(1,1)$ generano $\mathbb R^2$ (già solo i primi due bastano), ma sono dipendenti: $(1,1)=(1,0)+(0,1)$. Possiamo eliminare $(1,1)$ e restare con $(1,0),(0,1)$, che generano ancora tutto $\mathbb R^2$.

### Verso il numero minimo di generatori

Questo procedimento di eliminazione — togliere un vettore ridondante alla volta finché quelli rimasti non sono più linearmente dipendenti — si può ripetere. Il punto d'arrivo è un insieme di generatori che sono **anche** linearmente indipendenti: un insieme del genere non può essere ridotto ulteriormente (se togliessimo un vettore, non genereremmo più tutto $V$, perché quel vettore non era ridondante) né ha bisogno di essere ampliato (genera già tutto $V$). Questo insieme "minimo ma sufficiente" è esattamente ciò che chiameremo **base** nel prossimo notebook.

````python
import sympy as sp

# Verifica che v1=(1,0), v2=(0,1), v3=(1,1) generano R^2, e che v3 è ridondante
l1, l2, l3, u1, u2 = sp.symbols('lambda1 lambda2 lambda3 u1 u2')
v1, v2, v3 = sp.Matrix([1, 0]), sp.Matrix([0, 1]), sp.Matrix([1, 1])

# Un vettore generico u=(u1,u2): risolviamo con tutti e 3 i vettori (sistema sottodeterminato)
u = sp.Matrix([u1, u2])
eq = l1 * v1 + l2 * v2 + l3 * v3 - u
sol = sp.solve([eq[0], eq[1]], [l1, l2], dict=True)  # risolviamo per l1,l2 in funzione di l3
print("Con tutti e 3 i generatori, soluzione generale (l3 libero):", sol)

# v3 è combinazione di v1,v2?
sol_ridondanza = sp.solve([(v3 - l1*v1 - l2*v2)[0], (v3 - l1*v1 - l2*v2)[1]], [l1, l2])
print("v3 = l1*v1 + l2*v2 con:", sol_ridondanza)

# Solo v1, v2: generano R^2 per ogni (u1,u2)?
eq2 = l1 * v1 + l2 * v2 - u
sol2 = sp.solve([eq2[0], eq2[1]], [l1, l2])
print("Con solo v1, v2, soluzione per u generico:", sol2)
````

## Esercizio svolto e commentato

**Problema:** $v_1=(1,2)$, $v_2=(2,4)$, $v_3=(0,1)$ generano $\mathbb R^2$? Se sì, trova un sottoinsieme minimo che genera ancora tutto $\mathbb R^2$.

**Ragionamento:**

**Passo 1 — controlliamo prima le dipendenze**, prima di verificare la generazione: notiamo che $v_2=(2,4)=2(1,2)=2v_1$. Quindi $v_1,v_2,v_3$ sono linearmente dipendenti (come nel notebook 6.6): $v_2$ è ridondante.

**Passo 2 — eliminiamo $v_2$ e verifichiamo se $v_1,v_3$ generano $\mathbb R^2$.** Impostiamo, per un generico $u=(u_1,u_2)$: $\lambda_1(1,2)+\lambda_3(0,1)=(u_1,u_2)$, cioè $\begin{cases}\lambda_1=u_1\\ 2\lambda_1+\lambda_3=u_2\end{cases}$.

**Passo 3 — risolviamo**: dalla prima equazione $\lambda_1=u_1$; sostituendo nella seconda, $\lambda_3=u_2-2u_1$. Per **ogni** $(u_1,u_2)$ troviamo sempre una soluzione (nessuna condizione di compatibilità da verificare, a differenza dell'esercizio del notebook 6.5, perché qui abbiamo 2 equazioni e 2 incognite, non 3 equazioni e 2 incognite).

**Conclusione:** $v_1,v_3$ da soli generano già tutto $\mathbb R^2$, quindi anche $v_1,v_2,v_3$ (l'insieme più grande, che li contiene) generano $\mathbb R^2$. Un sottoinsieme minimo di generatori è $\{v_1,v_3\}=\{(1,2),(0,1)\}$: due vettori, linearmente indipendenti (verificabile con il criterio del notebook 6.6), che bastano da soli.

**Perché conviene sempre eliminare le dipendenze prima di cercare i generatori minimi:** il vettore $v_2$, essendo un multiplo di $v_1$, non poteva in nessun modo aiutare a "raggiungere" direzioni nuove — la sua rimozione era garantita non ridurre lo span, per il ragionamento generale fatto nella teoria. Riconoscere subito le dipendenze (spesso visibili "a occhio", come qui) evita di dover risolvere sistemi inutilmente complicati con più vettori del necessario.

````python
import sympy as sp

l1, l3, u1, u2 = sp.symbols('lambda1 lambda3 u1 u2')
v1, v3 = sp.Matrix([1, 2]), sp.Matrix([0, 1])
u = sp.Matrix([u1, u2])

eq = l1 * v1 + l3 * v3 - u
sol = sp.solve([eq[0], eq[1]], [l1, l3])
print("Soluzione per u=(u1,u2) generico:", sol)
print("Esiste sempre soluzione (per ogni u1,u2)? Sì, il sistema è determinato senza condizioni extra.")
````

## Riepilogo finale

- Un insieme di vettori **genera** $V$ se il loro span è tutto $V$: ogni vettore di $V$ è raggiungibile come combinazione lineare di essi.
- Se un insieme di generatori è linearmente dipendente, si può sempre eliminare un vettore ridondante senza cambiare lo span.
- Ripetendo l'eliminazione fino a esaurire le dipendenze, si arriva a un insieme di generatori **anche** linearmente indipendenti: minimo ma sufficiente.
- Per generare $\mathbb R^n$, servono **almeno** $n$ vettori indipendenti (intuizione anticipata negli esercizi, formalizzata nel prossimo notebook).
