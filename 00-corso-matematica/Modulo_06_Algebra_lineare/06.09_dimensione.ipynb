# 6.9 — Dimensione

**Prerequisiti:** 6.8 Basi.


## Perché ci interessa

Nel notebook 6.8 abbiamo visto due basi diverse di $\mathbb R^2$: $\{(1,0),(0,1)\}$ e $\{(1,1),(1,-1)\}$. Entrambe hanno **due** elementi. Non è un caso: intuitivamente, "il piano ha due direzioni indipendenti", e questo fatto dovrebbe restare vero qualunque base scegliamo. In questo notebook dimostriamo che è proprio così — tutte le basi di uno spazio vettoriale hanno lo stesso numero di elementi — e chiamiamo questo numero la **dimensione** dello spazio: un unico numero che cattura "quanto è grande" lo spazio in senso algebrico, indipendentemente da come lo descriviamo.

## Teoria, passo dopo passo

### Lemma di scambio (Steinitz)

Il risultato chiave, da cui discende tutto il resto, è questo:

> **Lemma.** Se $\{v_1,\dots,v_n\}$ genera $V$ e $\{w_1,\dots,w_m\}$ è un insieme di vettori di $V$ linearmente indipendenti, allora $m\le n$.

**Idea della dimostrazione (per sostituzione).** Poiché $v_1,\dots,v_n$ generano $V$, possiamo scrivere $w_1$ come loro combinazione lineare; poiché $w_1\ne\mathbf 0$ (essendo parte di un insieme indipendente), almeno uno dei coefficienti è non nullo, diciamo quello di $v_1$: possiamo allora "sostituire" $v_1$ con $w_1$, ottenendo un nuovo insieme di generatori $\{w_1,v_2,\dots,v_n\}$ (si dimostra che genera ancora $V$, risolvendo per $v_1$ in funzione di $w_1,v_2,\dots,v_n$ e sostituendo). Ripetendo il procedimento con $w_2$ (che si scrive come combinazione di $w_1,v_2,\dots,v_n$, e — usando l'indipendenza di $w_1,w_2$ — deve avere un coefficiente non nullo su uno dei $v_i$ rimanenti, altrimenti $w_2$ sarebbe combinazione del solo $w_1$, contraddicendo l'indipendenza di $w_1,w_2$) sostituiamo un altro $v_i$ con $w_2$. Se fosse $m>n$, dopo $n$ sostituzioni avremmo esaurito tutti i $v_i$, sostituendoli con $w_1,\dots,w_n$; ma resterebbe ancora $w_{n+1}$ da esprimere, che dovrebbe essere combinazione lineare di $w_1,\dots,w_n$ (gli unici generatori rimasti) — contraddicendo l'indipendenza dell'intero insieme $\{w_1,\dots,w_m\}$. Quindi deve essere $m\le n$.

### Teorema: tutte le basi hanno la stessa cardinalità

> **Teorema.** Se $\mathcal B_1=\{v_1,\dots,v_n\}$ e $\mathcal B_2=\{w_1,\dots,w_m\}$ sono due basi dello stesso spazio vettoriale $V$, allora $n=m$.

**Dimostrazione.** $\mathcal B_1$ genera $V$ e $\mathcal B_2$ è indipendente (essendo una base): per il lemma, $m\le n$. Simmetricamente, $\mathcal B_2$ genera $V$ e $\mathcal B_1$ è indipendente: per il lemma, $n\le m$. Da $m\le n$ e $n\le m$ segue $n=m$. $\blacksquare$

### Definizione di dimensione

Grazie a questo teorema, il numero di elementi di una base è un **invariante** dello spazio vettoriale: non dipende dalla base scelta. Possiamo quindi definire la **dimensione** di $V$, indicata $\dim V$, come il numero di elementi di una (qualsiasi) sua base.

- $\dim\mathbb R^n=n$ (la base canonica ha $n$ elementi).
- $\dim\{\mathbf 0\}=0$ (per convenzione: l'insieme vuoto è "base" del sottospazio banale, che non ha elementi non nulli).
- $\dim(\mathbb R^2[x])=3$ (i polinomi di grado $\le2$: una base è $\{1,x,x^2\}$, verificabile con lo stesso criterio usato per i vettori).

### Conseguenza pratica: contare invece di verificare due volte

Un corollario molto usato: se $\dim V=n$ ed abbiamo esattamente $n$ vettori $v_1,\dots,v_n\in V$, allora basta verificare **una sola** delle due condizioni (indipendenza oppure generazione) per concludere che formano una base — l'altra è automatica. (Intuizione: se $n$ vettori sono indipendenti in uno spazio di dimensione $n$, non c'è "spazio" per aggiungerne altri restando indipendenti, quindi devono già generare tutto; analogamente se $n$ vettori generano uno spazio di dimensione $n$, non se ne può togliere nessuno restando generatori, quindi devono essere indipendenti.) Questo dimezza il lavoro nella pratica.

````python
import sympy as sp

# Verifica: dim(R^2[x]) = 3, con base {1, x, x^2}
x = sp.symbols('x')
a0, a1, a2, l0, l1, l2 = sp.symbols('a0 a1 a2 lambda0 lambda1 lambda2')

p = a0 + a1*x + a2*x**2  # polinomio generico di grado <= 2
base = [sp.Integer(1), x, x**2]

# Generazione: ogni p si scrive come a0*1 + a1*x + a2*x^2 (banale per costruzione)
print("p =", p, " -> coefficienti rispetto a {1,x,x^2}: (a0,a1,a2) =", (a0, a1, a2))

# Indipendenza: l0*1 + l1*x + l2*x^2 = 0 (come polinomio) implica l0=l1=l2=0
combinazione = l0*1 + l1*x + l2*x**2
poly = sp.Poly(combinazione, x)
coeffs = poly.all_coeffs()
print("Coefficienti del polinomio combinazione (devono essere tutti nulli perché sia il polinomio nullo):", coeffs)

# Esempio: 4 vettori in R^3 (dim 3) devono essere dipendenti (lemma di Steinitz)
v1, v2, v3, v4 = sp.Matrix([1,0,0]), sp.Matrix([0,1,0]), sp.Matrix([0,0,1]), sp.Matrix([1,1,1])
c1,c2,c3,c4 = sp.symbols('c1 c2 c3 c4')
eq = c1*v1 + c2*v2 + c3*v3 + c4*v4
sol = sp.solve([eq[0], eq[1], eq[2]], [c1, c2, c3], dict=True)
print("\n4 vettori in R^3, soluzione (c4 libero, conferma dipendenza):", sol)
````

## Esercizio svolto e commentato

**Problema:** verifica, usando la scorciatoia vista sopra, se $v_1=(1,2,1)$, $v_2=(0,1,1)$, $v_3=(2,1,-1)$ formano una base di $\mathbb R^3$.

**Ragionamento:**

**Passo 1 — contiamo i vettori: sono 3, e $\dim\mathbb R^3=3$.** Per la scorciatoia vista nella teoria, basta verificare **una sola** delle due condizioni (generazione o indipendenza): scegliamo di verificare l'indipendenza, di solito più semplice da controllare direttamente.

**Passo 2 — impostiamo l'equazione omogenea**: $\lambda_1(1,2,1)+\lambda_2(0,1,1)+\lambda_3(2,1,-1)=(0,0,0)$, cioè:

$$\begin{cases}\lambda_1+2\lambda_3=0\\ 2\lambda_1+\lambda_2+\lambda_3=0\\ \lambda_1+\lambda_2-\lambda_3=0\end{cases}$$

**Passo 3 — dalla prima equazione**: $\lambda_1=-2\lambda_3$.

**Passo 4 — sostituiamo nella terza equazione**: $-2\lambda_3+\lambda_2-\lambda_3=0$, cioè $\lambda_2=3\lambda_3$.

**Passo 5 — verifichiamo con la seconda equazione**: $2(-2\lambda_3)+3\lambda_3+\lambda_3=-4\lambda_3+3\lambda_3+\lambda_3=0$. Sempre vera, qualunque $\lambda_3$ — a prima vista sembrerebbe un parametro libero...

**Passo 6 — ma attenzione: verifichiamo se $\lambda_3=0$ è l'**unica** possibilità compatibile con tutte e tre.** Rileggendo: abbiamo trovato $\lambda_1=-2\lambda_3$ e $\lambda_2=3\lambda_3$ per **ogni** $\lambda_3$, e la terza equazione era automaticamente soddisfatta. Questo significa che $\lambda_3$ è **davvero libero**: per esempio $\lambda_3=1$ dà $\lambda_1=-2,\lambda_2=3,\lambda_3=1$, non tutti nulli.

**Conclusione:** i tre vettori sono in realtà **linearmente dipendenti**, non formano una base. (Verifichiamo: $-2v_1+3v_2+v_3=-2(1,2,1)+3(0,1,1)+(2,1,-1)=(-2,-4,-2)+(0,3,3)+(2,1,-1)=(0,0,0)$. ✓)

**Perché questo esempio è istruttivo:** mostra che la scorciatoia (contare i vettori e verificare una sola condizione) **funziona correttamente solo se si verifica con attenzione** l'unica condizione scelta — qui, verificando l'indipendenza fino in fondo, abbiamo scoperto che $\lambda_3$ restava libero, e quindi i vettori **non** sono indipendenti (nonostante fossero esattamente 3, quanto la dimensione): il numero giusto di vettori non garantisce automaticamente che siano una base, bisogna comunque verificare la condizione scelta correttamente.

````python
import sympy as sp

l1, l2, l3 = sp.symbols('lambda1 lambda2 lambda3')
v1 = sp.Matrix([1, 2, 1])
v2 = sp.Matrix([0, 1, 1])
v3 = sp.Matrix([2, 1, -1])

eq = l1 * v1 + l2 * v2 + l3 * v3
soluzione = sp.solve([eq[0], eq[1], eq[2]], [l1, l2, l3])
print("Soluzione sistema omogeneo:", soluzione)  # se ha un parametro libero, sono dipendenti

# Verifica diretta della combinazione trovata
combinazione = -2*v1 + 3*v2 + 1*v3
print("Verifica -2v1+3v2+v3 =", combinazione.T)
````

## Riepilogo finale

- Il **lemma di Steinitz** garantisce che, se un insieme genera $V$ e un altro è indipendente in $V$, il secondo non può avere più elementi del primo.
- Da qui segue che **tutte le basi** di uno stesso spazio vettoriale hanno lo stesso numero di elementi: questo numero è la **dimensione** $\dim V$.
- $\dim\mathbb R^n=n$; $\dim\{\mathbf 0\}=0$; la dimensione di un sottospazio proprio di $\mathbb R^n$ è sempre minore di $n$.
- **Scorciatoia pratica**: con esattamente $\dim V$ vettori, basta verificare indipendenza **oppure** generazione (non entrambe) per concludere che formano una base — ma bisogna verificare la condizione scelta fino in fondo, con attenzione.
