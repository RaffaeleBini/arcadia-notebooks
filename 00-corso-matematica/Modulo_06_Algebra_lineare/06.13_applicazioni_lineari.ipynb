# 6.13 — Applicazioni lineari

**Prerequisiti:** 6.10 Matrici, 3.1 Concetto di funzione, 6.3 Spazi vettoriali.


## Perché ci interessa

Nel notebook 3.1 abbiamo definito una funzione come una regola che associa a ogni elemento di un insieme un elemento di un altro. Tra tutte le possibili funzioni tra spazi vettoriali, ce n'è una classe speciale — quelle che "rispettano" somma e prodotto per scalare, cioè che trasformano una combinazione lineare di vettori nella stessa combinazione lineare delle immagini. Sono le **applicazioni lineari**: rotazioni, riflessioni, proiezioni, scalature — trasformazioni geometriche che preservano la struttura vettoriale. Sono onnipresenti: dalla computer grafica (ogni rotazione di un modello 3D è un'applicazione lineare) alla meccanica quantistica.

## Teoria, passo dopo passo

### Definizione

Siano $V,W$ spazi vettoriali. Una funzione $T:V\to W$ è un'**applicazione lineare** (o trasformazione lineare) se, per ogni $u,v\in V$ e $\lambda\in\mathbb R$:

1. $T(u+v)=T(u)+T(v)$ (rispetta la somma);
2. $T(\lambda v)=\lambda T(v)$ (rispetta il prodotto per scalare).

Le due condizioni si possono riassumere in una sola: $T(\lambda_1v_1+\lambda_2v_2)=\lambda_1T(v_1)+\lambda_2T(v_2)$ per ogni $v_1,v_2\in V$ e $\lambda_1,\lambda_2\in\mathbb R$ — e più in generale, $T$ trasforma **ogni** combinazione lineare (notebook 6.5) in "la stessa" combinazione lineare delle immagini.

### Conseguenza immediata: $T(\mathbf 0)=\mathbf 0$

Da $T(\lambda v)=\lambda T(v)$, ponendo $\lambda=0$: $T(\mathbf 0)=T(0\cdot v)=0\cdot T(v)=\mathbf 0$. Ogni applicazione lineare manda il vettore nullo nel vettore nullo — un test rapido per **escludere** che una funzione sia lineare: se $T(\mathbf 0)\ne\mathbf 0$, $T$ non è lineare.

### Il prodotto matrice-vettore è un'applicazione lineare

Data una matrice $A$ ($m\times n$), la funzione $T(x)=Ax$ (notebook 6.10) è lineare. **Verifica:** $T(u+v)=A(u+v)=Au+Av=T(u)+T(v)$ (distributiva del prodotto matrice-vettore rispetto alla somma, verificabile componente per componente); $T(\lambda v)=A(\lambda v)=\lambda(Av)=\lambda T(v)$. Come vedremo nel prossimo notebook, vale anche il viceversa: **ogni** applicazione lineare tra spazi di dimensione finita si può scrivere come moltiplicazione per una matrice — è il motivo per cui abbiamo studiato le matrici prima delle applicazioni lineari.

### Esempi geometrici in $\mathbb R^2$

- **Rotazione** di angolo $\theta$ attorno all'origine: $T(x,y)=(x\cos\theta-y\sin\theta,\ x\sin\theta+y\cos\theta)$ (userà le identità trigonometriche del notebook 2.10 per verificarne la linearità — è lineare perché scritta come $Av$ con una matrice esplicita, che troveremo nel notebook 6.14).
- **Riflessione** rispetto all'asse $x$: $T(x,y)=(x,-y)$.
- **Scalatura**: $T(x,y)=(2x,2y)$ (raddoppia ogni vettore) — o, più in generale, $T(v)=\lambda v$.
- **Proiezione** sull'asse $x$: $T(x,y)=(x,0)$.

### Un esempio di funzione NON lineare

$T(x,y)=(x+1,y)$ (una traslazione) **non** è lineare: $T(\mathbf 0)=T(0,0)=(1,0)\ne(0,0)$, viola già il test rapido visto sopra. Intuitivamente, le traslazioni "spostano" l'origine, mentre ogni applicazione lineare deve necessariamente fissarla.

````python
import numpy as np

def T(v, A):
    return A @ v

# Verifica linearità del prodotto matrice-vettore, per una matrice A a caso
A = np.array([[2, 1], [0, 3]])
u = np.array([1, 2])
v = np.array([3, -1])
lam = 5

lhs = T(u + v, A)
rhs = T(u, A) + T(v, A)
print("T(u+v) =", lhs, " | T(u)+T(v) =", rhs, " -> uguali?", np.array_equal(lhs, rhs))

lhs2 = T(lam * v, A)
rhs2 = lam * T(v, A)
print("T(lambda*v) =", lhs2, " | lambda*T(v) =", rhs2, " -> uguali?", np.array_equal(lhs2, rhs2))

# T(0) per la rotazione di 90 gradi (lineare) e per una traslazione (non lineare)
theta = np.pi / 2
R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])
print("\nRotazione di 0:", R @ np.array([0, 0]), "-> lineare (manda 0 in 0)")

def traslazione(v):
    return v + np.array([1, 0])

print("Traslazione di 0:", traslazione(np.array([0, 0])), "-> NON lineare (0 non va in 0)")
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $T(x,y)=(x^2,y)$ è un'applicazione lineare.

**Ragionamento:**

**Passo 1 — applichiamo il test rapido: $T(\mathbf 0)$?** $T(0,0)=(0^2,0)=(0,0)$: il test non esclude la linearità (ma non la conferma nemmeno — è solo una condizione necessaria, non sufficiente).

**Passo 2 — verifichiamo direttamente la prima condizione**, $T(u+v)\overset?=T(u)+T(v)$, con un esempio concreto: prendiamo $u=(1,0)$, $v=(1,0)$. Allora $u+v=(2,0)$, e $T(u+v)=T(2,0)=(4,0)$.

**Passo 3 — calcoliamo l'altro membro**: $T(u)=T(1,0)=(1,0)$, $T(v)=T(1,0)=(1,0)$, quindi $T(u)+T(v)=(2,0)$.

**Passo 4 — confrontiamo**: $T(u+v)=(4,0)$ ma $T(u)+T(v)=(2,0)$: **non coincidono** ($4\ne2$).

**Conclusione:** $T$ **non è lineare**: abbiamo trovato un controesempio esplicito che viola la prima condizione della definizione.

**Perché basta un solo controesempio per escludere la linearità (mentre servirebbe una dimostrazione generale per confermarla):** la definizione richiede che $T(u+v)=T(u)+T(v)$ valga per **tutti** i vettori $u,v$; per negare un'affermazione universale (notebook 0.17, quantificatore $\forall$) basta un solo controesempio, mentre per **confermarla** servirebbe verificarla per una coppia $u,v$ generica (con componenti simboliche, come faremo di regola quando dimostriamo che una trasformazione è lineare). In questo caso, il termine $x^2$ è il segnale visivo che tradisce la non linearità: le applicazioni lineari, come vedremo nel prossimo notebook, sono sempre espresse da combinazioni lineari delle componenti in ingresso — mai potenze o prodotti tra variabili.

````python
import sympy as sp

x1, y1, x2, y2 = sp.symbols('x1 y1 x2 y2')

def T(x, y):
    return (x**2, y)

Tu = T(x1, y1)
Tv = T(x2, y2)
T_somma = T(x1 + x2, y1 + y2)

lhs = sp.expand(T_somma[0])
rhs = sp.expand(Tu[0] + Tv[0])
print("T(u+v)[0] =", lhs)
print("T(u)[0]+T(v)[0] =", rhs)
print("Uguali (in generale)?", sp.simplify(lhs - rhs) == 0)
````

## Riepilogo finale

- Un'**applicazione lineare** $T:V\to W$ rispetta somma e prodotto per scalare: $T(u+v)=T(u)+T(v)$ e $T(\lambda v)=\lambda T(v)$.
- Conseguenza necessaria (ma non sufficiente): $T(\mathbf 0)=\mathbf 0$ — un test rapido per escludere la linearità.
- Il prodotto matrice-vettore $T(x)=Ax$ è sempre un'applicazione lineare; come vedremo, è anche l'**unico** tipo possibile, in dimensione finita.
- Rotazioni, riflessioni, scalature e proiezioni sono applicazioni lineari; le traslazioni e le funzioni con prodotti/potenze delle variabili, in generale, no.
