## Teoria, passo dopo passo

### Forma matriciale di un sistema lineare

Un sistema di $m$ equazioni in $n$ incognite $x_1,\dots,x_n$:

$$\begin{cases}a_{11}x_1+\cdots+a_{1n}x_n=b_1\\ \vdots\\ a_{m1}x_1+\cdots+a_{mn}x_n=b_m\end{cases}$$

si scrive $Ax=b$, dove $A$ è la matrice $m\times n$ dei coefficienti e $b=(b_1,\dots,b_m)$ è il vettore dei termini noti (notebook 6.10). Per lavorare sul sistema, conviene affiancare a $A$ la colonna $b$, ottenendo la **matrice completa** $(A|b)$.

### L'idea dell'eliminazione di Gauss

L'eliminazione di Gauss generalizza il metodo di riduzione già visto nel notebook 1.7: si sommano multipli di una riga alle altre per **annullare progressivamente** i coefficienti sotto la diagonale, fino a ottenere un sistema "a scala" (triangolare), facile da risolvere per sostituzione all'indietro.

Le operazioni ammesse (dette **operazioni elementari di riga**) sono tre, e non cambiano l'insieme delle soluzioni del sistema:

1. scambiare due righe;
2. moltiplicare una riga per uno scalare non nullo;
3. sommare a una riga un multiplo di un'altra riga.

**Perché queste operazioni non cambiano le soluzioni:** ogni operazione elementare corrisponde a manipolare in modo lecito le equazioni originarie (esattamente come nel notebook 1.7, sommare o sottrarre equazioni tra loro, o moltiplicarle per una costante non nulla): un vettore $x$ soddisfa il sistema originale se e solo se soddisfa il sistema trasformato, perché ogni operazione è reversibile (si può sempre "tornare indietro" con l'operazione inversa).

### Un esempio guidato

Risolviamo $\begin{cases}x+y+z=6\\2x-y+z=3\\x+2y-z=2\end{cases}$. Matrice completa:

$$\left(\begin{array}{ccc|c}1&1&1&6\\2&-1&1&3\\1&2&-1&2\end{array}\right)$$

**Passo 1**: eliminiamo la $x$ dalla riga 2 e dalla riga 3, sottraendo multipli della riga 1. $R_2\to R_2-2R_1$: $(0,-3,-1,-9)$. $R_3\to R_3-R_1$: $(0,1,-2,-4)$.

$$\left(\begin{array}{ccc|c}1&1&1&6\\0&-3&-1&-9\\0&1&-2&-4\end{array}\right)$$

**Passo 2**: eliminiamo la $y$ dalla riga 3, usando la riga 2. $R_3\to R_3+\frac13R_2$: $\left(0,\,1-\frac13\times3,\,-2-\frac13,\,-4-3\right)=(0,0,-\frac73,-7)$.

$$\left(\begin{array}{ccc|c}1&1&1&6\\0&-3&-1&-9\\0&0&-\frac73&-7\end{array}\right)$$

**Passo 3 — sostituzione all'indietro.** Dall'ultima riga: $-\frac73z=-7\Rightarrow z=3$. Dalla seconda riga: $-3y-3=-9\Rightarrow y=2$. Dalla prima riga: $x+2+3=6\Rightarrow x=1$.

**Soluzione:** $(x,y,z)=(1,2,3)$.

### Rango e classificazione delle soluzioni

Il numero di righe non nulle nella forma "a scala" (dopo l'eliminazione) si chiama **rango** della matrice, indicato $\operatorname{rank}(A)$ — un concetto che riprenderemo a fondo nel notebook 6.15. Confrontando $\operatorname{rank}(A)$ con $\operatorname{rank}(A|b)$ e col numero di incognite $n$, si classifica il sistema (**teorema di Rouché-Capelli**):

- se $\operatorname{rank}(A)\ne\operatorname{rank}(A|b)$: il sistema è **incompatibile** (nessuna soluzione) — capita quando l'eliminazione produce una riga del tipo $(0,0,\dots,0\,|\,c)$ con $c\ne0$, un'equazione assurda "$0=c$";
- se $\operatorname{rank}(A)=\operatorname{rank}(A|b)=n$: soluzione **unica**;
- se $\operatorname{rank}(A)=\operatorname{rank}(A|b)<n$: **infinite soluzioni**, con $n-\operatorname{rank}(A)$ parametri liberi (è quello che è successo nell'esercizio del notebook 6.9, dove un $\lambda_3$ restava libero).

````python
import numpy as np
import sympy as sp

A = sp.Matrix([[1, 1, 1], [2, -1, 1], [1, 2, -1]])
b = sp.Matrix([6, 3, 2])

# Matrice completa e forma a scala (row echelon form)
completa = A.row_join(b)
scala, pivots = completa.rref()  # forma ridotta a scala
print("Matrice a scala ridotta (A|b):\n", scala)

# Soluzione diretta con SymPy
soluzione = A.solve(b)
print("\nSoluzione (x,y,z):", soluzione.T)

# Rango di A e di (A|b)
print("\nrango(A) =", A.rank(), " | rango(A|b) =", completa.rank())
````

## Esercizio svolto e commentato

**Problema:** risolvi (o classifica) il sistema $\begin{cases}x+2y=4\\2x+4y=9\end{cases}$.

**Ragionamento:**

**Passo 1 — impostiamo l'eliminazione**: matrice completa $\left(\begin{array}{cc|c}1&2&4\\2&4&9\end{array}\right)$.

**Passo 2 — eliminiamo la $x$ dalla riga 2**: $R_2\to R_2-2R_1$: $(2-2\times1,\,4-2\times2,\,9-2\times4)=(0,0,1)$.

$$\left(\begin{array}{cc|c}1&2&4\\0&0&1\end{array}\right)$$

**Passo 3 — interpretiamo l'ultima riga.** La riga $(0,0,1)$ corrisponde all'equazione $0\cdot x+0\cdot y=1$, cioè $0=1$: un'affermazione **falsa**, indipendentemente dai valori di $x,y$.

**Conclusione:** il sistema è **incompatibile** (nessuna soluzione). Per il criterio del rango: $\operatorname{rank}(A)=1$ (una sola riga non nulla nella parte dei coefficienti) ma $\operatorname{rank}(A|b)=2$ (la riga $(0,0,1)$ è non nulla anche includendo il termine noto): $\operatorname{rank}(A)\ne\operatorname{rank}(A|b)$, coerente col teorema di Rouché-Capelli.

**Perché questo risultato ha senso geometricamente:** le due equazioni rappresentano due rette nel piano (notebook 2.2); riscrivendole in forma esplicita, $y=-\frac12x+2$ e $y=-\frac12x+\frac94$, hanno lo **stesso coefficiente angolare** ($-\frac12$) ma diverso termine noto: sono rette **parallele e distinte**, che non si intersecano mai — coerente con l'assenza di soluzioni.

````python
import sympy as sp

A = sp.Matrix([[1, 2], [2, 4]])
b = sp.Matrix([4, 9])

completa = A.row_join(b)
scala, pivots = completa.rref()
print("Forma a scala:\n", scala)

print("\nrango(A) =", A.rank(), " | rango(A|b) =", completa.rank())
print("Compatibile?", A.rank() == completa.rank())
````

## Riepilogo finale

- Un sistema lineare $Ax=b$ si risolve con l'**eliminazione di Gauss**: operazioni elementari di riga che riducono la matrice completa $(A|b)$ a forma "a scala", senza cambiare l'insieme delle soluzioni.
- Il **rango** di una matrice è il numero di righe non nulle dopo l'eliminazione.
- Il **teorema di Rouché-Capelli** classifica il sistema confrontando $\operatorname{rank}(A)$, $\operatorname{rank}(A|b)$ e il numero di incognite $n$: incompatibile, unica soluzione, o infinite soluzioni.
- L'eliminazione di Gauss generalizza a $n$ incognite qualsiasi i metodi elementari del notebook 1.7, ed è alla base di quasi tutti gli algoritmi numerici di algebra lineare.
