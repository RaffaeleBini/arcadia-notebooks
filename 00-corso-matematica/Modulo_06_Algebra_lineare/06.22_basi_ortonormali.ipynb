# 6.22 — Basi ortonormali

**Prerequisiti:** 6.21 Ortogonalità, 6.16 Matrici invertibili.


## Perché ci interessa

Nell'esercizio 4 del notebook 6.21 abbiamo "normalizzato" una base ortogonale, dividendo ogni vettore per la sua norma, ottenendo una **base ortonormale**. Questo passo aggiuntivo (rispetto alla sola ortogonalità) non è cosmetico: rende ancora più semplice la formula delle coordinate, ed è collegato a un tipo speciale di matrice — la **matrice ortogonale** — che rappresenta esattamente le trasformazioni che preservano lunghezze e angoli (rotazioni e riflessioni). Sono la base concettuale, per esempio, della computer grafica 3D e della meccanica quantistica.

## Teoria, passo dopo passo

### Definizione

Una base $\{e_1,\dots,e_n\}$ di uno spazio euclideo $V$ (notebook 6.20) è **ortonormale** se è ortogonale (notebook 6.21) e, in più, ogni vettore ha norma $1$:

$$\langle e_i,e_j\rangle = \begin{cases}1 & \text{se } i=j\\ 0 & \text{se } i\ne j\end{cases}$$

(questa condizione compatta si indica spesso con il simbolo $\delta_{ij}$, il **delta di Kronecker**).

### Coordinate ancora più semplici

Per una base ortonormale, la formula delle coordinate del notebook 6.21 si semplifica ulteriormente: poiché $\langle e_i,e_i\rangle=1$, non serve più dividere:

$$\lambda_i = \langle u,e_i\rangle$$

Le coordinate di $u$ rispetto a una base ortonormale sono semplicemente i prodotti scalari di $u$ con ciascun vettore di base — il calcolo più diretto possibile.

### Matrici ortogonali

Una matrice quadrata $Q$ ($n\times n$) è **ortogonale** se le sue colonne formano una base ortonormale di $\mathbb R^n$ (con il prodotto scalare standard). Una conseguenza notevole:

$$Q^TQ = I \qquad\text{cioè}\qquad Q^{-1}=Q^T$$

**Perché:** l'elemento $(i,j)$ di $Q^TQ$ è, per definizione di prodotto matriciale (notebook 6.10), il prodotto scalare tra la riga $i$ di $Q^T$ (cioè la colonna $i$ di $Q$) e la colonna $j$ di $Q$: cioè $\langle q_i,q_j\rangle$, dove $q_i,q_j$ sono le colonne di $Q$. Per ortonormalità, questo vale $1$ se $i=j$ e $0$ altrimenti: esattamente gli elementi della matrice identità. Quindi calcolare l'inversa di una matrice ortogonale è immediato — basta **trasporla**, senza calcoli con determinanti o minori (notebook 6.16).

### Le matrici ortogonali preservano lunghezze e angoli

> **Fatto.** Se $Q$ è ortogonale, allora $\|Qv\|=\|v\|$ per ogni $v$ (preserva le lunghezze), e $\langle Qv,Qw\rangle=\langle v,w\rangle$ (preserva i prodotti scalari, quindi gli angoli).

**Dimostrazione (lunghezze).** $\|Qv\|^2=\langle Qv,Qv\rangle=(Qv)^T(Qv)=v^TQ^TQv=v^TIv=v^Tv=\langle v,v\rangle=\|v\|^2$ (usando $Q^TQ=I$). Estraendo la radice: $\|Qv\|=\|v\|$.

Questo conferma quanto anticipato nel notebook 6.14: rotazioni e riflessioni (le trasformazioni che "non deformano" le figure) sono esattamente le trasformazioni date da matrici ortogonali — la matrice di rotazione $\begin{pmatrix}\cos\theta&-\sin\theta\\\sin\theta&\cos\theta\end{pmatrix}$ è ortogonale, come verificheremo.

````python
import sympy as sp

theta = sp.symbols('theta', real=True)
Q = sp.Matrix([[sp.cos(theta), -sp.sin(theta)], [sp.sin(theta), sp.cos(theta)]])

# Verifica Q^T Q = I
prodotto = sp.simplify(Q.T * Q)
print("Q^T * Q =\n", prodotto)

# Verifica che le colonne sono ortonormali
col1, col2 = Q[:, 0], Q[:, 1]
print("\n||col1||^2 =", sp.simplify(col1.dot(col1)))
print("||col2||^2 =", sp.simplify(col2.dot(col2)))
print("col1 . col2 =", sp.simplify(col1.dot(col2)))

# Verifica che preserva la norma di un vettore generico
x, y = sp.symbols('x y', real=True)
v = sp.Matrix([x, y])
Qv = Q * v
norma_v = sp.sqrt(v.dot(v))
norma_Qv = sp.sqrt(sp.simplify(Qv.dot(Qv)))
print("\n||v|| =", norma_v)
print("||Qv|| =", norma_Qv)
````

## Esercizio svolto e commentato

**Problema:** verifica che $Q=\begin{pmatrix}\frac{\sqrt2}2&-\frac{\sqrt2}2\\\frac{\sqrt2}2&\frac{\sqrt2}2\end{pmatrix}$ è ortogonale, poi usala per trovare le coordinate di $u=(3,1)$ rispetto alla base ortonormale data dalle sue colonne.

**Ragionamento:**

**Passo 1 — verifichiamo che le colonne hanno norma 1**: colonna 1, $\left(\frac{\sqrt2}2\right)^2+\left(\frac{\sqrt2}2\right)^2=\frac24+\frac24=1$. ✓ Analogamente per la colonna 2 (stessi valori, solo un segno diverso, che non cambia il quadrato).

**Passo 2 — verifichiamo l'ortogonalità delle colonne**: $\frac{\sqrt2}2\times\left(-\frac{\sqrt2}2\right)+\frac{\sqrt2}2\times\frac{\sqrt2}2=-\frac24+\frac24=0$. ✓

**Passo 3 — concludiamo che $Q$ è ortogonale** (entrambe le condizioni verificate; riconosciamo anche che $Q$ è la matrice di rotazione con $\theta=45°$, dato che $\cos45°=\sin45°=\frac{\sqrt2}2$).

**Passo 4 — calcoliamo le coordinate di $u=(3,1)$ usando la formula semplificata** $\lambda_i=\langle u,e_i\rangle$ (nessuna divisione necessaria, essendo la base ortonormale): $\lambda_1=\langle u,e_1\rangle=3\times\frac{\sqrt2}2+1\times\frac{\sqrt2}2=\frac{4\sqrt2}2=2\sqrt2$. $\lambda_2=\langle u,e_2\rangle=3\times\left(-\frac{\sqrt2}2\right)+1\times\frac{\sqrt2}2=\frac{-2\sqrt2}2=-\sqrt2$.

**Verifica**: $2\sqrt2\,e_1-\sqrt2\,e_2=2\sqrt2\left(\frac{\sqrt2}2,\frac{\sqrt2}2\right)-\sqrt2\left(-\frac{\sqrt2}2,\frac{\sqrt2}2\right)=(2,2)-(-1,1)=(3,1)=u$. ✓

**Perché, essendo $Q$ ortogonale, potremmo anche trovare le coordinate calcolando semplicemente $Q^Tu$ invece di due prodotti scalari separati:** dato che $Q^{-1}=Q^T$ per le matrici ortogonali (proprietà vista nella teoria), e le coordinate di $u$ rispetto alla base data dalle colonne di $Q$ sono per definizione $x=Q^{-1}u$ (risolvendo $Qx=u$, come nel notebook 6.16), otteniamo $x=Q^Tu$ — un singolo prodotto matrice-vettore, invece di due prodotti scalari calcolati separatamente: la stessa informazione, organizzata in modo ancora più compatto.

````python
import sympy as sp

s2 = sp.sqrt(2) / 2
Q = sp.Matrix([[s2, -s2], [s2, s2]])
u = sp.Matrix([3, 1])

print("Q^T Q =\n", sp.simplify(Q.T * Q))

coordinate_diretta = sp.Matrix([u.dot(Q[:,0]), u.dot(Q[:,1])])
coordinate_QT = Q.T * u
print("\nCoordinate (prodotti scalari):", coordinate_diretta.T)
print("Coordinate (Q^T u):", sp.simplify(coordinate_QT).T)
````

4. *(più impegnativo)* Dimostra che il determinante di una matrice ortogonale è sempre $\pm1$ (usa $Q^TQ=I$ e le proprietà del determinante del notebook 6.12 e 6.19).

<details>
<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** Colonne $(0,1)$ e $(1,0)$: entrambe norma $1$ ($0^2+1^2=1$, $1^2+0^2=1$), prodotto scalare $0\times1+1\times0=0$. **È ortogonale.**

**2.** Colonne $(1,1)$ e $(1,-1)$: norma di $(1,1)$ è $\sqrt2\ne1$. **Non è ortogonale** (nonostante le colonne siano ortogonali tra loro: $1\times1+1\times(-1)=0$). Per correggerla, basta normalizzare ogni colonna dividendola per la sua norma: $Q'=\begin{pmatrix}\frac1{\sqrt2}&\frac1{\sqrt2}\\\frac1{\sqrt2}&-\frac1{\sqrt2}\end{pmatrix}$.

**3.** Per matrici ortogonali, $Q^{-1}=Q^T$: basta trasporre. $Q^T=\begin{pmatrix}0&1\\1&0\end{pmatrix}$ (in questo caso, $Q$ è simmetrica, quindi $Q^{-1}=Q$ stessa).

**4.** Da $Q^TQ=I$, prendendo il determinante di entrambi i membri: $\det(Q^TQ)=\det I=1$. Per la proprietà del determinante del prodotto (notebook 6.12): $\det(Q^T)\det(Q)=1$. Usando $\det(Q^T)=\det Q$ (proprietà vista nel notebook 6.12): $(\det Q)^2=1$, quindi $\det Q=\pm1$.

Verifica computazionale:

Q2 = sp.Matrix([[1,1],[1,-1]])
print("Es.2:", Q2.T * Q2)

print("Es.4, det(Q1):", Q1.det())

## Riepilogo finale

- Una base **ortonormale** è ortogonale e ha tutti i vettori di norma $1$; le coordinate si calcolano con la formula più semplice possibile: $\lambda_i=\langle u,e_i\rangle$.
- Una **matrice ortogonale** $Q$ ha colonne ortonormali, e soddisfa $Q^TQ=I$, cioè $Q^{-1}=Q^T$: l'inversa si calcola per semplice trasposizione.
- Le matrici ortogonali preservano lunghezze e angoli: rappresentano rotazioni e riflessioni.
- Il determinante di una matrice ortogonale è sempre $\pm1$.
