# 6.17 — Autovalori

**Prerequisiti:** 6.14 Matrice associata, 6.12 Determinanti, 1.6 Equazioni di secondo grado.


## Perché ci interessa

Nel notebook 6.14 abbiamo visto che una matrice $2\times2$ può rappresentare rotazioni, riflessioni, scalature — trasformazioni spesso complicate da visualizzare in generale. Ma per molte matrici esistono **direzioni speciali**: vettori che, applicando $A$, non cambiano direzione — vengono solo "allungati" o "accorciati" (eventualmente ribaltati) di un fattore fisso. Trovare queste direzioni privilegiate è enormemente utile: se una trasformazione agisce semplicemente come una scalatura lungo certe direzioni, possiamo capirne il comportamento (e calcolare le sue potenze, utile per esempio nello studio di sistemi dinamici) in modo molto più semplice che manipolando la matrice generica.

## Teoria, passo dopo passo

### Definizione

Sia $A$ una matrice quadrata $n\times n$. Un numero $\lambda\in\mathbb R$ (o $\mathbb C$, ma qui restiamo nel caso reale) è un **autovalore** di $A$ se esiste un vettore $v\ne\mathbf 0$ tale che:

$$Av = \lambda v$$

Il vettore $v$ si chiama **autovettore** associato a $\lambda$ (lo studieremo a fondo nel prossimo notebook): applicando $A$, $v$ non cambia direzione, viene solo scalato di un fattore $\lambda$.

### Come trovare gli autovalori: l'equazione caratteristica

Riscriviamo $Av=\lambda v$ come $Av-\lambda v=\mathbf 0$, cioè $(A-\lambda I)v=\mathbf 0$ (dove $I$ è la matrice identità, notebook 6.16, necessaria per sottrarre $\lambda$ da una matrice — non ha senso sottrarre uno scalare direttamente da una matrice). Questa è un'equazione omogenea (notebook 6.6): per avere una soluzione $v\ne\mathbf 0$, la matrice $A-\lambda I$ **non** deve essere invertibile (altrimenti, per il notebook 6.16, l'unica soluzione sarebbe $v=\mathbf 0$, che escludiamo per definizione di autovettore). Per il criterio del determinante (notebook 6.16):

$$\det(A-\lambda I) = 0$$

Questa equazione (nell'incognita $\lambda$) si chiama **equazione caratteristica**; il polinomio $p(\lambda)=\det(A-\lambda I)$ si chiama **polinomio caratteristico**. Gli autovalori di $A$ sono esattamente le radici di questo polinomio.

### Calcolo per una matrice $2\times2$

Per $A=\begin{pmatrix}a&b\\c&d\end{pmatrix}$: $A-\lambda I=\begin{pmatrix}a-\lambda&b\\c&d-\lambda\end{pmatrix}$, quindi $\det(A-\lambda I)=(a-\lambda)(d-\lambda)-bc=\lambda^2-(a+d)\lambda+(ad-bc)$: un'**equazione di secondo grado** (notebook 1.6) in $\lambda$, con al più due soluzioni reali. Notiamo che $a+d$ è la **traccia** di $A$ (somma della diagonale) e $ad-bc=\det A$: il polinomio caratteristico $2\times2$ è sempre $\lambda^2-\operatorname{tr}(A)\lambda+\det A$.

### Esempio guidato

$A=\begin{pmatrix}3&1\\0&2\end{pmatrix}$: $\operatorname{tr}(A)=5$, $\det A=6$. Polinomio caratteristico: $\lambda^2-5\lambda+6=0$. Con la formula risolutiva (notebook 1.6): $\lambda=\frac{5\pm\sqrt{25-24}}2=\frac{5\pm1}2$, cioè $\lambda_1=3$, $\lambda_2=2$.

### Un fatto notevole: matrici triangolari

Se $A$ è **triangolare** (tutti gli elementi sopra o sotto la diagonale sono nulli, come l'esempio appena visto), gli autovalori sono semplicemente gli elementi della diagonale. Lo si vede perché $A-\lambda I$ resta triangolare, e il determinante di una matrice triangolare è il prodotto della diagonale (generalizzazione dello sviluppo di Laplace ripetuto lungo la prima colonna o riga, che a ogni passo elimina tutti i termini tranne quello diagonale): $\det(A-\lambda I)=(a_{11}-\lambda)(a_{22}-\lambda)\cdots(a_{nn}-\lambda)$, che si annulla esattamente per $\lambda=a_{ii}$.

````python
import sympy as sp

lam = sp.symbols('lambda')
A = sp.Matrix([[3, 1], [0, 2]])

# Polinomio caratteristico
I = sp.eye(2)
p = (A - lam * I).det()
print("Polinomio caratteristico:", sp.expand(p))

autovalori = sp.solve(p, lam)
print("Autovalori:", autovalori)

# Confronto con il metodo diretto di SymPy
print("Autovalori (metodo diretto):", A.eigenvals())

# Verifica: traccia e determinante
print("\ntraccia(A) =", A.trace(), " | somma autovalori =", sum(autovalori))
print("det(A) =", A.det(), " | prodotto autovalori =", autovalori[0] * autovalori[1])
````

## Esercizio svolto e commentato

**Problema:** trova gli autovalori di $A=\begin{pmatrix}2&-1\\1&2\end{pmatrix}$.

**Ragionamento:**

**Passo 1 — calcoliamo traccia e determinante**: $\operatorname{tr}(A)=2+2=4$, $\det A=2\times2-(-1)\times1=4+1=5$.

**Passo 2 — scriviamo il polinomio caratteristico**: $\lambda^2-4\lambda+5=0$.

**Passo 3 — applichiamo la formula risolutiva** (notebook 1.6): $\lambda=\frac{4\pm\sqrt{16-20}}2=\frac{4\pm\sqrt{-4}}2$.

**Passo 4 — interpretiamo il discriminante negativo.** $\sqrt{-4}$ non è un numero reale: il discriminante $\Delta=16-20=-4<0$ significa che, restando nei numeri reali, **questa matrice non ha autovalori reali**.

**Conclusione:** $A$ non ha autovalori reali (avrebbe due autovalori complessi coniugati, $\lambda=2\pm i$, se lavorassimo con i numeri complessi, che introdurremo nel Modulo 9).

**Perché questo risultato ha senso geometricamente:** ricordando l'esercizio 3 del notebook 6.14, una matrice del tipo $\begin{pmatrix}\cos\theta&-\sin\theta\\\sin\theta&\cos\theta\end{pmatrix}$ rappresenta una rotazione; la nostra $A$ ha una struttura simile (elementi diagonali uguali, elementi fuori diagonale opposti in segno) — ed è proprio una rotazione (composta con una scalatura). Una rotazione (diversa da $0°$ o $180°$) **non può** avere autovalori reali: non esiste nessun vettore non nullo che, ruotato, mantenga la sua direzione — ogni vettore cambia direzione, per definizione di rotazione. L'assenza di autovalori reali è quindi coerente, anzi prevedibile, dalla natura geometrica della trasformazione.

````python
import sympy as sp

lam = sp.symbols('lambda')
A = sp.Matrix([[2, -1], [1, 2]])

I = sp.eye(2)
p = (A - lam * I).det()
print("Polinomio caratteristico:", sp.expand(p))

discriminante = A.trace()**2 - 4 * A.det()
print("Discriminante:", discriminante)

print("Autovalori (SymPy, in campo complesso):", A.eigenvals())
````

## Riepilogo finale

- Un **autovalore** $\lambda$ di $A$ è uno scalare per cui esiste $v\ne\mathbf 0$ con $Av=\lambda v$: una direzione che $A$ preserva, scalandola.
- Gli autovalori sono le radici del **polinomio caratteristico** $\det(A-\lambda I)=0$.
- Per $2\times2$: $\lambda^2-\operatorname{tr}(A)\lambda+\det A=0$; per matrici triangolari, gli autovalori sono gli elementi diagonali.
- Non tutte le matrici reali hanno autovalori reali: le rotazioni (diverse da $0°,180°$) ne sono un esempio, coerentemente col fatto che nessuna direzione resta invariata.
