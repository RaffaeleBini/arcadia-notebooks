# 6.4 — Sottospazi

**Prerequisiti:** 6.3 Spazi vettoriali.


## Perché ci interessa

Nell'esercizio svolto del notebook 6.3 abbiamo verificato che $V=\{(x,y,z):x+y+z=0\}$ è uno spazio vettoriale, ma per farlo abbiamo dovuto controllare solo la chiusura rispetto alle operazioni e l'esistenza di neutro e opposti — non tutti e 8 gli assiomi, perché $V$ era già "dentro" $\mathbb R^3$, di cui ereditava gratuitamente le proprietà algebriche di base. Questa scorciatoia è così utile e così comune (capita ogni volta che vogliamo studiare "una parte" di uno spazio vettoriale già noto) che merita un nome proprio e un criterio preciso: il concetto di **sottospazio**.

## Teoria, passo dopo passo

### Definizione

Sia $V$ uno spazio vettoriale. Un sottoinsieme $W\subseteq V$ è un **sottospazio** di $V$ se $W$, con le stesse operazioni di $V$ ristrette a $W$, è a sua volta uno spazio vettoriale.

### Il criterio del sottospazio

Verificare tutti e 8 gli assiomi ogni volta sarebbe ripetitivo: come intuito nel notebook 6.3, gli assiomi "strutturali" (associativa, commutativa, distributive, ecc.) valgono automaticamente in $W$ perché valgono già in tutto $V$. Basta quindi controllare tre condizioni, molto più rapide da verificare:

> $W\subseteq V$ è un sottospazio se e solo se:
> 1. $\mathbf 0\in W$ (il vettore nullo appartiene a $W$);
> 2. per ogni $v,w\in W$, anche $v+w\in W$ (chiusura rispetto alla somma);
> 3. per ogni $v\in W$ e $\lambda\in\mathbb R$, anche $\lambda v\in W$ (chiusura rispetto al prodotto per scalare).

**Perché queste tre condizioni bastano:** gli assiomi 1, 2, 5, 6, 7, 8 dello spazio vettoriale (notebook 6.3) sono identità che coinvolgono solo il "comportamento" delle operazioni (es. $v+w=w+v$), non l'appartenenza a un insieme particolare — se sono vere per ogni elemento di $V$, restano vere in particolare per gli elementi di $W\subseteq V$. Restano da garantire solo l'esistenza del neutro (condizione 1) e dell'opposto per ogni elemento: ma se $W$ è chiuso rispetto al prodotto per scalare (condizione 3), allora per ogni $v\in W$ anche $(-1)v=-v\in W$, quindi l'esistenza degli opposti è già garantita dalla condizione 3 e non serve verificarla a parte.

**Nota pratica:** se $\mathbf 0\notin W$, possiamo concludere **subito** che $W$ non è un sottospazio, senza controllare altro (era esattamente il problema dell'esercizio 2 del notebook 6.3, la retta $y=x+1$ che non passa per l'origine).

### Esempi geometrici in $\mathbb R^3$

- $\{\mathbf 0\}$ (il solo vettore nullo) è sempre un sottospazio: il **sottospazio banale**.
- Una retta passante per l'origine è un sottospazio (di dimensione 1, concetto che formalizzeremo nel notebook 6.9).
- Un piano passante per l'origine è un sottospazio (di dimensione 2).
- $V$ stesso è un sottospazio di sé stesso.
- Una retta o un piano che **non** passa per l'origine **non** è mai un sottospazio (condizione 1 violata).

### Un controesempio: un insieme che non è un sottospazio

Consideriamo $W=\{(x,y)\in\mathbb R^2 : xy\ge0\}$ (primo e terzo quadrante, assi inclusi). Contiene $\mathbf 0=(0,0)$ (condizione 1 ok). Ma non è chiuso rispetto alla somma: prendiamo $v=(2,1)\in W$ (primo quadrante, $xy=2\ge0$) e $w=(-1,-3)\in W$ (terzo quadrante, $xy=3\ge0$). La somma è $v+w=(1,-2)$, che ha $xy=-2<0$: **non** appartiene a $W$. Quindi $W$ non è chiuso rispetto alla somma: non è un sottospazio (è l'unione di due "coni" opposti, non un oggetto "piatto" come una retta o un piano).

````python
import sympy as sp

# Verifica: W = {(x,y,z) : x + y + z = 0} è chiuso rispetto a somma e prodotto per scalare
x1, y1, z1, x2, y2, z2, lam = sp.symbols('x1 y1 z1 x2 y2 z2 lambda')

# v = (x1,y1,z1) con z1 = -x1-y1 (per stare in W)
z1_w = -x1 - y1
z2_w = -x2 - y2

somma_ok = sp.simplify((x1 + x2) + (y1 + y2) + (z1_w + z2_w)) == 0
scalare_ok = sp.simplify(lam * x1 + lam * y1 + lam * z1_w) == 0
print("Chiuso rispetto alla somma?", somma_ok)
print("Chiuso rispetto al prodotto per scalare?", scalare_ok)
print("Contiene il vettore nullo? (0,0,0):", 0 + 0 + 0 == 0)

# Controesempio: W = {(x,y): x*y >= 0} NON è un sottospazio
v = (2, 1)
w = (-1, -3)
somma = (v[0] + w[0], v[1] + w[1])
print("\nv =", v, "xy =", v[0] * v[1], ">= 0?", v[0]*v[1] >= 0)
print("w =", w, "xy =", w[0] * w[1], ">= 0?", w[0]*w[1] >= 0)
print("v+w =", somma, "xy =", somma[0] * somma[1], ">= 0?", somma[0]*somma[1] >= 0)
````

## Esercizio svolto e commentato

**Problema:** stabilisci se $W=\{p(x)\in\mathbb R^2[x] : p(1)=0\}$ (i polinomi di grado $\le2$ che si annullano in $x=1$) è un sottospazio di $\mathbb R^2[x]$.

**Ragionamento:**

**Passo 1 — verifichiamo la condizione 1: il vettore nullo appartiene a $W$?** Il polinomio nullo $p(x)=0$ soddisfa $p(1)=0$ banalmente. Sì, $\mathbf 0\in W$.

**Passo 2 — verifichiamo la condizione 2: chiusura rispetto alla somma.** Siano $p,q\in W$, cioè $p(1)=0$ e $q(1)=0$. Consideriamo $(p+q)(x)=p(x)+q(x)$: valutando in $x=1$, $(p+q)(1)=p(1)+q(1)=0+0=0$ (qui usiamo che valutare una somma di polinomi in un punto equivale a sommare i valori, proprietà della somma di funzioni). Quindi $p+q\in W$.

**Passo 3 — verifichiamo la condizione 3: chiusura rispetto al prodotto per scalare.** Sia $p\in W$ e $\lambda\in\mathbb R$: $(\lambda p)(1)=\lambda\, p(1)=\lambda\times0=0$. Quindi $\lambda p\in W$.

**Conclusione:** tutte e tre le condizioni sono soddisfatte: $W$ è un sottospazio di $\mathbb R^2[x]$.

**Perché conviene ragionare per "valutazione in un punto" invece che sui coefficienti:** si potrebbe scrivere $p(x)=a_0+a_1x+a_2x^2$ con la condizione $p(1)=a_0+a_1+a_2=0$ e ripetere lo stesso ragionamento sui coefficienti — arriveremmo alla stessa conclusione, ma il ragionamento per valutazione è più diretto: la condizione "$p(1)=0$" è **lineare** nei coefficienti (è una somma pesata di essi, uguagliata a zero), e ogni condizione lineare omogenea di questo tipo definisce automaticamente un sottospazio — un fatto generale che ritroveremo studiando i sistemi lineari omogenei (notebook 6.11).

````python
import sympy as sp

x = sp.symbols('x')
a0, a1, a2, b0, b1, b2, lam = sp.symbols('a0 a1 a2 b0 b1 b2 lambda')

p = a0 + a1 * x + a2 * x**2
q = b0 + b1 * x + b2 * x**2

# Imponiamo p(1)=0 e q(1)=0, cioè a0+a1+a2=0 e b0+b1+b2=0
a0_w = -a1 - a2
b0_w = -b1 - b2

p_w = p.subs(a0, a0_w)
q_w = q.subs(b0, b0_w)

somma = sp.expand(p_w + q_w)
print("(p+q)(1) =", somma.subs(x, 1))       # deve fare 0

scalato = sp.expand(lam * p_w)
print("(lambda*p)(1) =", sp.simplify(scalato.subs(x, 1)))  # deve fare 0
````

## Riepilogo finale

- Un **sottospazio** $W$ di uno spazio vettoriale $V$ è un sottoinsieme che è a sua volta uno spazio vettoriale con le stesse operazioni.
- **Criterio pratico**: basta verificare che $\mathbf 0\in W$, che $W$ sia chiuso rispetto alla somma, e che sia chiuso rispetto al prodotto per scalare — non serve riverificare tutti gli 8 assiomi.
- Se $\mathbf 0\notin W$, si può concludere subito che $W$ non è un sottospazio.
- Geometricamente in $\mathbb R^2$/$\mathbb R^3$: i sottospazi sono rette e piani passanti per l'origine (oltre al sottospazio banale $\{\mathbf 0\}$ e a $V$ stesso) — insiemi "piatti" e illimitati, mai regioni curve o limitate come dischi o coni.
