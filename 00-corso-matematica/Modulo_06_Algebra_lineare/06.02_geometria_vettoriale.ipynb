# 6.2 — Geometria vettoriale

**Prerequisiti:** 6.1 Vettori, 2.3 Distanza tra punti, 2.9 Seno, coseno, tangente.


## Perché ci interessa

Nel notebook 6.1 abbiamo imparato a sommare vettori e a moltiplicarli per uno scalare, ma non abbiamo ancora detto **quanto è lungo** un vettore, né come misurare l'**angolo** tra due vettori. Sono domande naturali: se $v$ rappresenta uno spostamento, la sua lunghezza è la distanza percorsa; se $v$ e $w$ rappresentano due forze, l'angolo tra loro determina come si combinano. In questo notebook introduciamo due strumenti — la **norma** e il **prodotto scalare** — che rispondono a queste domande e che saranno centrali per tutto il resto del modulo (ortogonalità, proiezioni, basi ortonormali).

## Teoria, passo dopo passo

### La norma (lunghezza) di un vettore

Per un vettore $v=(v_1,v_2)$ nel piano, la sua lunghezza è la distanza dall'origine $(0,0)$ al punto $(v_1,v_2)$, che sappiamo calcolare con la formula della distanza (notebook 2.3): $d=\sqrt{v_1^2+v_2^2}$. Chiamiamo questa quantità **norma** di $v$:

$$\|v\| = \sqrt{v_1^2+v_2^2}$$

In $\mathbb R^n$, generalizzando: $\|v\|=\sqrt{v_1^2+v_2^2+\cdots+v_n^2}$. Un vettore con $\|v\|=1$ si dice **versore** (o vettore unitario); dato $v\ne\mathbf 0$, il vettore $v/\|v\|$ è sempre un versore con la stessa direzione di $v$ (verifichiamolo: la sua norma è $\|v\|/\|v\|=1$).

### Il prodotto scalare (o "prodotto punto")

Dati $v=(v_1,v_2)$ e $w=(w_1,w_2)$, il **prodotto scalare** è definito così:

$$v\cdot w = v_1w_1+v_2w_2$$

(in $\mathbb R^n$: $v\cdot w=\sum_{i=1}^n v_iw_i$). A differenza della somma di vettori, il risultato è un **numero** (uno scalare), non un vettore — da qui il nome. Notiamo subito che $v\cdot v=v_1^2+v_2^2=\|v\|^2$: il prodotto scalare di un vettore con sé stesso è il quadrato della sua norma.

### Interpretazione geometrica: l'angolo tra due vettori

Il fatto sorprendente (che qui verifichiamo con un caso concreto, e giustifichiamo con il teorema del coseno) è che il prodotto scalare codifica anche l'**angolo** $\theta$ tra i due vettori:

$$v\cdot w = \|v\|\,\|w\|\,\cos\theta$$

**Perché vale questa formula:** consideriamo il triangolo con lati $v$, $w$ e $v-w$ (i tre vettori "chiudono" un triangolo, per la regola punta-coda del notebook 6.1). Il teorema del coseno applicato a questo triangolo (generalizzazione del teorema di Pitagora, che vedremo per esteso nel Modulo 2 sulla trigonometria) dà:

$$\|v-w\|^2 = \|v\|^2+\|w\|^2-2\|v\|\|w\|\cos\theta$$

D'altra parte, sviluppando $\|v-w\|^2=(v-w)\cdot(v-w)$ usando la distributiva del prodotto scalare (che verificheremo nella cella di codice) otteniamo $\|v\|^2+\|w\|^2-2\,v\cdot w$. Confrontando i due sviluppi di $\|v-w\|^2$: $-2\|v\|\|w\|\cos\theta=-2\,v\cdot w$, cioè proprio $v\cdot w=\|v\|\|w\|\cos\theta$.

### Ortogonalità

Dalla formula precedente segue un criterio semplicissimo: poiché $\cos(90°)=0$ (notebook 2.9), due vettori non nulli sono **ortogonali** (perpendicolari) se e solo se il loro prodotto scalare è zero:

$$v\perp w \iff v\cdot w=0$$

Questo criterio algebrico, che non richiede di disegnare nulla, sarà usato costantemente nei prossimi notebook.

### Proiezione di un vettore su un altro

Dato $w\ne\mathbf 0$, la **proiezione** di $v$ su $w$ è il vettore $\operatorname{proj}_w v$ che rappresenta "quanto di $v$ punta nella direzione di $w$". Si dimostra (scomponendo $v$ come somma di una componente parallela a $w$ e una ortogonale a $w$, e imponendo l'ortogonalità con il criterio appena visto) che:

$$\operatorname{proj}_w v = \frac{v\cdot w}{\|w\|^2}\,w$$

````python
import numpy as np

v = np.array([3, 4])
w = np.array([1, 2])

print("||v|| =", np.linalg.norm(v))          # sqrt(3^2+4^2) = 5
print("v . w =", np.dot(v, w))
print("||v|| * ||w|| * cos(theta) =", np.linalg.norm(v) * np.linalg.norm(w) * np.cos(
    np.arccos(np.dot(v, w) / (np.linalg.norm(v) * np.linalg.norm(w)))))

# Verifica distributiva del prodotto scalare: (v-w).(v-w) = ||v||^2 + ||w||^2 - 2 v.w
lhs = np.dot(v - w, v - w)
rhs = np.dot(v, v) + np.dot(w, w) - 2 * np.dot(v, w)
print("\n(v-w).(v-w) =", lhs, " | ||v||^2+||w||^2-2v.w =", rhs)

# Angolo tra v e w in gradi
cos_theta = np.dot(v, w) / (np.linalg.norm(v) * np.linalg.norm(w))
theta = np.degrees(np.arccos(cos_theta))
print("\nAngolo tra v e w:", theta, "gradi")

# Due vettori ortogonali: prodotto scalare nullo
a = np.array([1, 0])
b = np.array([0, 1])
print("\na . b =", np.dot(a, b), "-> ortogonali?", np.dot(a, b) == 0)

# Proiezione di v su w
proj = (np.dot(v, w) / np.dot(w, w)) * w
print("\nproiezione di v su w:", proj)
````

## Esercizio svolto e commentato

**Problema:** dati $v=(2,2)$ e $w=(3,-3)$, stabilisci se sono ortogonali. Se non lo sono, calcola l'angolo tra loro.

**Ragionamento:**

**Passo 1 — applichiamo il criterio di ortogonalità**: calcoliamo il prodotto scalare $v\cdot w=2\times3+2\times(-3)=6-6=0$.

**Passo 2 — interpretiamo il risultato.** Poiché $v\cdot w=0$ e nessuno dei due vettori è nullo, per il criterio visto sopra $v$ e $w$ sono **ortogonali**: non serve nemmeno calcolare l'angolo, sappiamo già che è $90°$.

**Verifica geometrica (per controllo):** $v=(2,2)$ punta lungo la bisettrice del primo quadrante (a $45°$ dall'asse $x$); $w=(3,-3)$ punta lungo la bisettrice del quarto quadrante (a $-45°$). La differenza tra le due direzioni è $45°-(-45°)=90°$: coerente con il calcolo algebrico.

**Perché conviene sempre usare il criterio algebrico invece di "ad occhio":** in questo esempio la simmetria rendeva visibile l'ortogonalità, ma in generale — con vettori in $\mathbb R^3$ o più dimensioni, dove non possiamo disegnare — il prodotto scalare è l'**unico** modo pratico per verificare la perpendicolarità: da qui l'importanza di questo criterio in tutto il seguito del modulo.

````python
import numpy as np

v = np.array([2, 2])
w = np.array([3, -3])

prodotto = np.dot(v, w)
print("v . w =", prodotto)
print("Ortogonali?", prodotto == 0)
````

## Riepilogo finale

- La **norma** $\|v\|=\sqrt{v_1^2+\dots+v_n^2}$ misura la lunghezza di un vettore; un versore ha norma $1$.
- Il **prodotto scalare** $v\cdot w=\sum v_iw_i$ è un numero (non un vettore) che codifica l'angolo tra $v$ e $w$: $v\cdot w=\|v\|\|w\|\cos\theta$.
- **Criterio di ortogonalità**: $v\perp w \iff v\cdot w=0$ — il modo pratico, algebrico, per verificare la perpendicolarità anche senza disegnare.
- La **proiezione** di $v$ su $w$ è $\operatorname{proj}_w v=\dfrac{v\cdot w}{\|w\|^2}w$: la "ombra" di $v$ nella direzione di $w$.
