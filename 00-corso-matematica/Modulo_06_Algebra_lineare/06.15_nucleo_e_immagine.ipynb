# 6.15 — Nucleo e immagine

**Prerequisiti:** 6.13 Applicazioni lineari, 6.14 Matrice associata, 3.5 Funzioni iniettive, suriettive, biiettive, 6.9 Dimensione.


## Perché ci interessa

Nel notebook 3.5 abbiamo definito iniettività e suriettività per funzioni generiche: una funzione è iniettiva se elementi diversi hanno immagini diverse, suriettiva se ogni elemento del codominio viene raggiunto. Per un'applicazione lineare $T:V\to W$, queste proprietà si possono studiare tramite due sottospazi molto concreti — il **nucleo** (o kernel) e l'**immagine** — che, come vedremo, sono legati da una relazione numerica elegante e potente: il **teorema della dimensione** (o teorema del rango).

## Teoria, passo dopo passo

### Il nucleo

Il **nucleo** di $T:V\to W$, indicato $\ker T$, è l'insieme dei vettori di $V$ che $T$ manda nel vettore nullo:

$$\ker T = \{v\in V : T(v)=\mathbf 0\}$$

**$\ker T$ è sempre un sottospazio di $V$.** Verifica (criterio del notebook 6.4): $T(\mathbf 0)=\mathbf 0$ (notebook 6.13), quindi $\mathbf 0\in\ker T$. Se $u,v\in\ker T$, cioè $T(u)=T(v)=\mathbf 0$, allora $T(u+v)=T(u)+T(v)=\mathbf 0+\mathbf 0=\mathbf 0$ (linearità), quindi $u+v\in\ker T$. Analogamente $T(\lambda v)=\lambda T(v)=\lambda\mathbf 0=\mathbf 0$, quindi $\lambda v\in\ker T$.

Se $T(x)=Ax$, allora $\ker T$ è esattamente l'insieme delle soluzioni del sistema omogeneo $Ax=\mathbf 0$ (notebook 6.11) — lo stesso sistema che abbiamo usato per verificare l'indipendenza lineare (notebook 6.6)!

### Il nucleo misura l'iniettività

> **Teorema.** $T$ è iniettiva se e solo se $\ker T=\{\mathbf 0\}$ (il nucleo contiene **solo** il vettore nullo).

**Dimostrazione.** ($\Rightarrow$) Se $T$ è iniettiva e $v\in\ker T$, cioè $T(v)=\mathbf 0=T(\mathbf 0)$, allora per iniettività $v=\mathbf 0$: quindi $\ker T=\{\mathbf 0\}$. ($\Leftarrow$) Supponiamo $\ker T=\{\mathbf 0\}$ e siano $u,v$ con $T(u)=T(v)$. Allora $T(u)-T(v)=\mathbf 0$, cioè (per linearità) $T(u-v)=\mathbf 0$: quindi $u-v\in\ker T=\{\mathbf 0\}$, cioè $u-v=\mathbf 0$, cioè $u=v$. Abbiamo mostrato $T(u)=T(v)\Rightarrow u=v$: è la definizione di iniettività (notebook 3.5). $\blacksquare$

### L'immagine

L'**immagine** di $T$, indicata $\operatorname{Im}T$, è l'insieme dei vettori di $W$ effettivamente raggiunti:

$$\operatorname{Im}T = \{T(v) : v\in V\} = \{w\in W : \exists v\in V,\ T(v)=w\}$$

$\operatorname{Im}T$ è un sottospazio di $W$ (verifica analoga a quella per $\ker T$, lasciata come esercizio). Se $T(x)=Ax$ e $v_1,\dots,v_n$ sono le colonne di $A$, allora $\operatorname{Im}T=\operatorname{span}(v_1,\dots,v_n)$ (notebook 6.5): l'immagine è esattamente lo span delle colonne. Per definizione, $T$ è **suriettiva** se e solo se $\operatorname{Im}T=W$.

### Il teorema della dimensione (rango-nullità)

> **Teorema.** Se $T:V\to W$ è lineare e $V$ ha dimensione finita, allora:
> $$\dim\ker T + \dim\operatorname{Im}T = \dim V$$

Il numero $\dim\ker T$ si chiama **nullità**, e $\dim\operatorname{Im}T$ si chiama **rango** di $T$ (coincide col rango della matrice associata, notebook 6.11). **Intuizione della dimostrazione:** si prende una base $\{u_1,\dots,u_k\}$ di $\ker T$ e la si estende a una base $\{u_1,\dots,u_k,v_1,\dots,v_r\}$ di tutto $V$ (con $k+r=\dim V$); si dimostra poi che $\{T(v_1),\dots,T(v_r)\}$ è una base di $\operatorname{Im}T$ — i vettori $u_i$ "scompaiono" (vanno in $\mathbf 0$), mentre i $v_j$ generano fedelmente l'immagine, senza ridondanze grazie all'iniettività "residua" fuori dal nucleo.

````python
import sympy as sp

# Applicazione T(x,y,z) = (x+y, y+z), matrice associata A
A = sp.Matrix([[1, 1, 0], [0, 1, 1]])

x, y, z = sp.symbols('x y z')
v = sp.Matrix([x, y, z])
print("T(x,y,z) =", (A * v).T)

# Nucleo: soluzioni di A@v = 0
nucleo = A.nullspace()
print("\nBase del nucleo:", [n.T for n in nucleo])
print("Dimensione del nucleo (nullità):", len(nucleo))

# Immagine: span delle colonne di A = colonne indipendenti
rango = A.rank()
print("\nRango di A (dimensione dell'immagine):", rango)

# Verifica teorema rango-nullità: dim(ker) + dim(Im) = dim(V) = 3 (V=R^3)
print("\nnullità + rango =", len(nucleo) + rango, " | dim(V) =", 3)
````

## Esercizio svolto e commentato

**Problema:** per $T:\mathbb R^3\to\mathbb R^2$ data da $A=\begin{pmatrix}1&2&3\\2&4&6\end{pmatrix}$, trova $\ker T$ e $\operatorname{Im}T$, e verifica il teorema della dimensione.

**Ragionamento:**

**Passo 1 — troviamo il nucleo, risolvendo $Ax=\mathbf 0$**: $\begin{cases}x+2y+3z=0\\2x+4y+6z=0\end{cases}$. Notiamo che la seconda equazione è $2\times$ la prima ($2x+4y+6z=2(x+2y+3z)$): in realtà è **un'unica equazione indipendente**, non due.

**Passo 2 — risolviamo l'unica equazione indipendente** $x+2y+3z=0$, cioè $x=-2y-3z$, con $y,z$ liberi. Il nucleo è $\ker T=\{(-2y-3z,\,y,\,z):y,z\in\mathbb R\}=\operatorname{span}((-2,1,0),\,(-3,0,1))$: ha **dimensione 2** (due parametri liberi, due generatori indipendenti — verificabile).

**Passo 3 — troviamo l'immagine**: $\operatorname{Im}T=\operatorname{span}((1,2),(2,4),(3,6))$ (le colonne di $A$). Ma $(2,4)=2(1,2)$ e $(3,6)=3(1,2)$: tutte e tre le colonne sono multipli di $(1,2)$! Quindi $\operatorname{Im}T=\operatorname{span}((1,2))$: ha **dimensione 1**.

**Passo 4 — verifichiamo il teorema della dimensione**: $\dim\ker T+\dim\operatorname{Im}T=2+1=3=\dim\mathbb R^3$. ✓

**Conclusione:** $T$ non è iniettiva (il nucleo non è $\{\mathbf 0\}$: ha dimensione 2, non 0) né suriettiva ($\operatorname{Im}T$ ha dimensione 1, non 2 come $\mathbb R^2$).

**Perché il fatto che le colonne di $A$ fossero tutte proporzionali "spiega" contemporaneamente sia il nucleo grande sia l'immagine piccola:** quando le colonne di $A$ sono fortemente dipendenti (qui, tutte multiple di una sola), lo span delle colonne — cioè $\operatorname{Im}T$ — collassa a una dimensione molto più bassa del numero di colonne; ma per il teorema della dimensione, se $\operatorname{Im}T$ "perde" dimensioni rispetto a $V$, quelle dimensioni **devono** ricomparire da qualche parte — e ricompaiono esattamente nel nucleo, che si allarga in proporzione. Nucleo e immagine sono due facce della stessa medaglia: quanta informazione l'applicazione "perde" (nucleo) e quanta ne "conserva" nel risultato (immagine), e la loro somma è vincolata a restare costante, pari a $\dim V$.

````python
import sympy as sp

A = sp.Matrix([[1, 2, 3], [2, 4, 6]])

nucleo = A.nullspace()
print("Base del nucleo:", [n.T for n in nucleo])
print("dim(ker T) =", len(nucleo))

rango = A.rank()
print("dim(Im T) = rango(A) =", rango)

print("Somma:", len(nucleo) + rango, " = dim(R^3) = 3 ?", len(nucleo) + rango == 3)
````

## Esercizi da risolvere

2. Per la stessa $A$ dell'esercizio 1, trova $\dim\operatorname{Im}T$ usando il teorema della dimensione (senza ricalcolare il rango da zero).
3. Per $T:\mathbb R^2\to\mathbb R^2$ data da $A=\begin{pmatrix}1&2\\2&4\end{pmatrix}$, trova $\ker T$ e $\operatorname{Im}T$.
4. *(più impegnativo)* Dimostra che, se $T:V\to W$ con $\dim V=\dim W$, allora $T$ è iniettiva se e solo se è suriettiva (usa il teorema della dimensione).

Quindi $T$ iniettiva $\iff T$ suriettiva.

## Riepilogo finale

- Il **nucleo** $\ker T=\{v:T(v)=\mathbf 0\}$ è un sottospazio di $V$; $T$ è iniettiva se e solo se $\ker T=\{\mathbf 0\}$.
- L'**immagine** $\operatorname{Im}T$ è un sottospazio di $W$, uguale allo span delle colonne della matrice associata; $T$ è suriettiva se e solo se $\operatorname{Im}T=W$.
- Il **teorema della dimensione**: $\dim\ker T+\dim\operatorname{Im}T=\dim V$ — nucleo e immagine si "compensano" sempre a sommare la dimensione di partenza.
- Se $\dim V=\dim W$, iniettività e suriettività diventano equivalenti.
