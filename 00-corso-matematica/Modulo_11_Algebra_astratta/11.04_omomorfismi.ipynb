## Perché ci interessa

Nel notebook 6.13 abbiamo studiato le **applicazioni lineari**: funzioni tra spazi vettoriali che rispettano somma e moltiplicazione per scalare, cioè $T(v+w)=T(v)+T(w)$. Questa idea — una funzione che "rispetta l'operazione" — si generalizza immediatamente ai gruppi: un **omomorfismo** è una funzione $f:G\to G'$ con $f(a*b)=f(a)*'f(b)$. Il nucleo (analogo al nucleo del notebook 6.15) misura "quanto" la funzione non è iniettiva, e collega direttamente — nel **primo teorema di isomorfismo**, il risultato centrale di questo notebook — il gruppo quoziente del notebook 11.3 con l'immagine dell'omomorfismo, mostrando che sono, in un senso preciso, "la stessa cosa".

## Teoria, passo dopo passo

### Definizione di omomorfismo

Una funzione $f:(G,*)\to(G',*')$ è un **omomorfismo di gruppi** se $f(a*b)=f(a)*'f(b)$ per ogni $a,b\in G$. Se $f$ è biiettiva, si chiama **isomorfismo**, e si scrive $G\cong G'$ ("$G$ e $G'$ sono isomorfi") — significa che $G$ e $G'$ sono **strutturalmente identici**, differendo solo nel nome degli elementi.

### Proprietà elementari

**Teorema.** Se $f:G\to G'$ è un omomorfismo, allora $f(e_G)=e_{G'}$ e $f(a^{-1})=f(a)^{-1}$ per ogni $a$.

**Dimostrazione.** $f(e_G)=f(e_G*e_G)=f(e_G)*'f(e_G)$ (applicando l'omomorfismo). Componendo entrambi i membri con $f(e_G)^{-1}$ (che esiste in $G'$, essendo $G'$ un gruppo): $e_{G'}=f(e_G)^{-1}*'f(e_G)=f(e_G)^{-1}*'(f(e_G)*'f(e_G))=(f(e_G)^{-1}*'f(e_G))*'f(e_G)=e_{G'}*'f(e_G)=f(e_G)$. Dunque $f(e_G)=e_{G'}$. Per l'inverso: $f(a)*'f(a^{-1})=f(a*a^{-1})=f(e_G)=e_{G'}$, quindi $f(a^{-1})$ soddisfa la definizione di inverso di $f(a)$; per l'unicità dell'inverso (notebook 11.1), $f(a^{-1})=f(a)^{-1}$. $\blacksquare$

### Nucleo e immagine

Il **nucleo** di $f$ è $\ker f=\{a\in G:f(a)=e_{G'}\}$; l'**immagine** è $\operatorname{Im}f=\{f(a):a\in G\}\subseteq G'$ — esattamente l'analogo, per i gruppi, del nucleo e dell'immagine di un'applicazione lineare (notebook 6.15).

**Teorema.** $\ker f\trianglelefteq G$ (è un sottogruppo **normale**), e $\operatorname{Im}f\le G'$ (è un sottogruppo).

**Dimostrazione (nucleo, sottogruppo).** $e_G\in\ker f$ (per la proprietà appena vista), quindi non vuoto. Dati $a,b\in\ker f$: $f(a*b^{-1})=f(a)*'f(b^{-1})=f(a)*'f(b)^{-1}=e_{G'}*'e_{G'}^{-1}=e_{G'}$, quindi $a*b^{-1}\in\ker f$: per il criterio del sottogruppo (notebook 11.2), $\ker f$ è un sottogruppo.

**Dimostrazione (nucleo, normale).** Per ogni $g\in G$ e $k\in\ker f$: $f(gkg^{-1})=f(g)*'f(k)*'f(g^{-1})=f(g)*'e_{G'}*'f(g)^{-1}=f(g)*'f(g)^{-1}=e_{G'}$, quindi $gkg^{-1}\in\ker f$: $g(\ker f)g^{-1}\subseteq\ker f$ per ogni $g$, e con lo stesso argomento applicato a $g^{-1}$ si ottiene l'inclusione opposta, dunque l'uguaglianza: $\ker f\trianglelefteq G$.

**Dimostrazione (immagine, sottogruppo).** $e_{G'}=f(e_G)\in\operatorname{Im}f$. Dati $f(a),f(b)\in\operatorname{Im}f$: $f(a)*'f(b)^{-1}=f(a)*'f(b^{-1})=f(a*b^{-1})\in\operatorname{Im}f$ (essendo $a*b^{-1}\in G$): per il criterio del sottogruppo, $\operatorname{Im}f\le G'$. $\blacksquare$

### Iniettività e nucleo banale

**Teorema.** $f$ è iniettiva $\iff$ $\ker f=\{e_G\}$.

**Dimostrazione ($\Rightarrow$).** Se $f$ è iniettiva e $a\in\ker f$: $f(a)=e_{G'}=f(e_G)$, quindi per iniettività $a=e_G$: $\ker f=\{e_G\}$.

**Dimostrazione ($\Leftarrow$).** Supponiamo $\ker f=\{e_G\}$ e $f(a)=f(b)$. Allora $f(a*b^{-1})=f(a)*'f(b)^{-1}=f(a)*'f(a)^{-1}=e_{G'}$ (usando $f(b)=f(a)$), quindi $a*b^{-1}\in\ker f=\{e_G\}$, cioè $a*b^{-1}=e_G$, cioè $a=b$. $\blacksquare$ Questo criterio (analogo a quello per le applicazioni lineari, notebook 6.15, dove iniettività equivale a nucleo banale) è enormemente più comodo della definizione diretta di iniettività, riducendo il controllo a un solo elemento anziché a tutte le coppie.

### Il primo teorema di isomorfismo

**Teorema.** Se $f:G\to G'$ è un omomorfismo, allora $G/\ker f\cong\operatorname{Im}f$.

**Dimostrazione (traccia).** Definiamo $\varphi:G/\ker f\to\operatorname{Im}f$ da $\varphi(a\ker f)=f(a)$. *Ben posta*: se $a\ker f=b\ker f$, allora $a^{-1}b\in\ker f$ (notebook 11.2), quindi $f(a^{-1}b)=e_{G'}$, cioè $f(a)^{-1}*'f(b)=e_{G'}$, cioè $f(a)=f(b)$: $\varphi$ non dipende dal rappresentante. *Omomorfismo*: $\varphi((a\ker f)*(b\ker f))=\varphi((ab)\ker f)=f(ab)=f(a)*'f(b)=\varphi(a\ker f)*'\varphi(b\ker f)$. *Iniettiva*: se $\varphi(a\ker f)=e_{G'}$, cioè $f(a)=e_{G'}$, allora $a\in\ker f$, cioè $a\ker f=\ker f$ è la classe neutra — per il criterio appena dimostrato, $\varphi$ ha nucleo banale, quindi è iniettiva. *Suriettiva su $\operatorname{Im}f$*: ogni $f(a)\in\operatorname{Im}f$ è raggiunto da $a\ker f$, per costruzione. Dunque $\varphi$ è un isomorfismo: $G/\ker f\cong\operatorname{Im}f$. $\blacksquare$

Questo teorema è il risultato centrale della teoria elementare dei gruppi: dice che **ogni** omomorfismo "è" essenzialmente una proiezione sul quoziente per il proprio nucleo, seguita da un'inclusione — la struttura di ogni omomorfismo è completamente determinata dal suo nucleo.

````python
from sympy.combinatorics import Permutation, PermutationGroup

# Esempio: f: Z -> Z/nZ, f(k) = k mod n, e' un omomorfismo con nucleo nZ
def f(k, n):
    return k % n

n = 5
print(f"Omomorfismo f: Z -> Z/{n}Z, f(k) = k mod {n}")
for a, b in [(3, 4), (7, -2), (10, 3)]:
    print(f"f({a}+{b}) = f({a+b}) = {f(a+b, n)},  f({a})+f({b}) mod {n} = {(f(a,n)+f(b,n)) % n},  uguali? "
          f"{f(a+b, n) == (f(a,n)+f(b,n)) % n}")

print(f"\nNucleo di f (multipli di {n} tra -20 e 20):", [k for k in range(-20, 21) if f(k, n) == 0])
print("Il nucleo e' esattamente 5Z, un sottogruppo normale di Z (Z e' abeliano).")
````

## Esercizio svolto e commentato

**Problema:** dimostra che l'applicazione determinante $\det:GL_n(\mathbb R)\to(\mathbb R\setminus\{0\},\times)$ è un omomorfismo di gruppi, trova il suo nucleo, e usa il primo teorema di isomorfismo per identificare $GL_n(\mathbb R)/SL_n(\mathbb R)$.

**Ragionamento:**

**Passo 1 — verifichiamo che è un omomorfismo.** Per la proprietà moltiplicativa del determinante (notebook 6.12), $\det(AB)=\det A\cdot\det B$ per ogni $A,B\in GL_n(\mathbb R)$: questa è esattamente la condizione di omomorfismo, con l'operazione di $GL_n(\mathbb R)$ (prodotto tra matrici) che corrisponde all'operazione di $(\mathbb R\setminus\{0\},\times)$ (prodotto tra numeri).

**Passo 2 — calcoliamo il nucleo.** $\ker(\det)=\{A\in GL_n(\mathbb R):\det A=1\}=SL_n(\mathbb R)$ — esattamente il sottogruppo delle matrici a determinante $1$ già incontrato nel notebook 11.2. Per il teorema di questo notebook, $SL_n(\mathbb R)\trianglelefteq GL_n(\mathbb R)$ **automaticamente**, senza bisogno di verificarlo a mano con la coniugazione (come avremmo dovuto fare usando solo la definizione).

**Passo 3 — calcoliamo l'immagine.** $\det$ è **suriettiva** su $\mathbb R\setminus\{0\}$: per ogni $c\ne0$, la matrice diagonale $\operatorname{diag}(c,1,\ldots,1)$ ha determinante $c$ (prodotto degli elementi diagonali, notebook 6.12), quindi $\operatorname{Im}(\det)=\mathbb R\setminus\{0\}$.

**Conclusione — applichiamo il primo teorema di isomorfismo.** $GL_n(\mathbb R)/SL_n(\mathbb R)\cong\operatorname{Im}(\det)=(\mathbb R\setminus\{0\},\times)$. **L'intuizione dietro questo risultato**: il quoziente "dimentica" tutto ciò che riguarda $SL_n(\mathbb R)$ (le matrici a determinante $1$, che diventano l'elemento neutro nel quoziente), lasciando visibile **solo** l'informazione contenuta nel valore del determinante — precisamente ciò che l'isomorfismo con $\mathbb R\setminus\{0\}$ afferma.

````python
import sympy as sp
import numpy as np

# Verifica numerica della proprieta' moltiplicativa del determinante
np.random.seed(0)
A = np.random.randint(1, 5, (3, 3))
B = np.random.randint(1, 5, (3, 3))

detA, detB, detAB = np.linalg.det(A), np.linalg.det(B), np.linalg.det(A @ B)
print("det(A) * det(B) =", detA * detB)
print("det(AB) =", detAB)
print("Coincidono (a meno di errori numerici)?", np.isclose(detA * detB, detAB))

# Suriettivita': costruiamo una matrice con un determinante c qualsiasi
c = 7.5
diag_c = np.diag([c, 1, 1])
print(f"\nMatrice diagonale con determinante richiesto c={c}: det =", np.linalg.det(diag_c))
````

## Riepilogo finale

- Un **omomorfismo** $f:G\to G'$ rispetta l'operazione: $f(a*b)=f(a)*'f(b)$; manda sempre il neutro nel neutro e gli inversi negli inversi.
- Il **nucleo** $\ker f$ è sempre un sottogruppo **normale**; l'**immagine** è sempre un sottogruppo; $f$ è iniettiva $\iff\ker f=\{e\}$.
- Il **primo teorema di isomorfismo** ($G/\ker f\cong\operatorname{Im}f$) è il risultato centrale: ogni omomorfismo è, essenzialmente, una proiezione sul quoziente seguita da un'inclusione.
- Applicato al determinante, dà $GL_n(\mathbb R)/SL_n(\mathbb R)\cong(\mathbb R\setminus\{0\},\times)$, mostrando la normalità di $SL_n(\mathbb R)$ senza calcoli diretti.
