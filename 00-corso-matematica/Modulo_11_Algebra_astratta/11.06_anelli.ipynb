## Perché ci interessa

I gruppi hanno una sola operazione. Ma $\mathbb Z$, le matrici quadrate (notebook 6.10), i polinomi (notebook 1.1) hanno **due** operazioni — somma e prodotto — legate dalla proprietà **distributiva** nominata per la prima volta nel notebook 0.5. Un **anello** è la struttura astratta che cattura esattamente questa situazione: un gruppo abeliano rispetto alla somma, un monoide (associativo, con neutro) rispetto al prodotto, con la distributiva a collegarle. Questo notebook apre la seconda metà dell'algebra astratta, dedicata a strutture con due operazioni — un percorso che, attraverso ideali (11.7) e domini integrali (11.9), condurrà ai campi (11.10) e infine alla teoria di Galois (11.13).

## Teoria, passo dopo passo

### Definizione di anello

Un **anello** $(R,+,\times)$ è un insieme $R$ con due operazioni tali che:

1. $(R,+)$ è un **gruppo abeliano** (neutro $0$, opposto $-a$ per ogni $a$);
2. $\times$ è **associativa**: $(a\times b)\times c=a\times(b\times c)$;
3. **Distributiva** (bilaterale): $a\times(b+c)=a\times b+a\times c$ e $(a+b)\times c=a\times c+b\times c$.

Se $\times$ ha un elemento neutro $1$ (con $1\ne0$), l'anello si dice **unitario**; se $\times$ è commutativa, l'anello si dice **commutativo**. Nota bene: a differenza dei gruppi, **non** si richiede che ogni elemento non nullo abbia un inverso moltiplicativo — è proprio l'assenza di questa richiesta a rendere gli anelli una struttura più generale dei campi (che vedremo nel notebook 11.10).

### Esempi fondamentali

- $(\mathbb Z,+,\times)$: anello commutativo unitario — l'esempio prototipo, ma **non** un gruppo rispetto a $\times$ (nessun intero eccetto $\pm1$ ha inverso moltiplicativo intero).
- $(\mathbb Z/n\mathbb Z,+,\times)$: anello commutativo unitario, con la moltiplicazione modulo $n$ (ben definita con lo stesso argomento usato per la somma nell'esercizio del notebook 11.1).
- Le matrici quadrate $n\times n$, $M_n(\mathbb R)$, con somma e prodotto tra matrici (notebook 6.10): anello unitario (neutro moltiplicativo la matrice identità) ma **non commutativo** per $n\ge2$.
- I polinomi $\mathbb R[x]$ (notebook 1.1), con somma e prodotto di polinomi: anello commutativo unitario — lo studieremo in dettaglio nel notebook 11.11.
- $2\mathbb Z$ (interi pari): anello commutativo **senza** elemento neutro moltiplicativo (nessun numero pari, moltiplicato per un altro numero pari, può dare un pari qualsiasi come "identità" — verificabile direttamente).

### Proprietà elementari: lo zero "assorbe" il prodotto

**Teorema.** In ogni anello, $a\times0=0\times a=0$ per ogni $a$.

**Dimostrazione.** $a\times0=a\times(0+0)=a\times0+a\times0$ (distributiva, usando $0+0=0$ nel gruppo additivo). Sottraendo $a\times0$ da entrambi i membri (cioè sommando l'opposto $-(a\times0)$, che esiste perché $(R,+)$ è un gruppo): $0=a\times0$. Analogamente per $0\times a$, usando la distributiva a destra. $\blacksquare$

Questa dimostrazione, di nuovo, usa **solo** gli assiomi: è sorprendente che un fatto così "ovvio" numericamente richieda una dimostrazione dagli assiomi, ma è esattamente questo il prezzo (e il vantaggio) dell'astrazione — il fatto vale automaticamente in **ogni** anello, anche i più esotici, perché discende dagli assiomi e non da proprietà specifiche dei numeri.

### La regola dei segni, dimostrata dagli assiomi

**Teorema.** In ogni anello, $(-a)\times b=-(a\times b)$ per ogni $a,b$.

**Dimostrazione.** $(-a)\times b+a\times b=((-a)+a)\times b$ (distributiva, raccogliendo $b$) $=0\times b=0$ (per il teorema precedente). Dunque $(-a)\times b$ è l'opposto di $a\times b$ nel gruppo additivo, cioè $(-a)\times b=-(a\times b)$ (per l'unicità dell'inverso in un gruppo, notebook 11.1, applicata al gruppo $(R,+)$). $\blacksquare$ Da qui segue, applicando il risultato due volte, $(-a)\times(-b)=-(a\times(-b))=-(-(a\times b))=a\times b$: la celebre "meno per meno fa più" del notebook 0.2, ora dimostrata **da zero**, per ogni anello, non solo per i numeri interi.

### Divisori dello zero

Un elemento $a\ne0$ di un anello è un **divisore dello zero** se esiste $b\ne0$ con $a\times b=0$. In $\mathbb Z$ non esistono divisori dello zero (se $ab=0$ con $a\ne0$, allora $b=0$ — un fatto familiare, dato per scontato ogni volta che si "semplifica" un'equazione), ma in $\mathbb Z/6\mathbb Z$: $2\times3=6\equiv0\pmod6$, con $2,3\ne0$ — sono entrambi divisori dello zero! Questa distinzione, apparentemente tecnica, è cruciale: la presenza di divisori dello zero è **esattamente** ciò che impedisce di "cancellare" un fattore comune in un'equazione, e sarà l'oggetto del notebook 11.9 (domini integrali, gli anelli senza questo problema).

````python
import numpy as np

# Verifica computazionale: lo zero assorbe il prodotto, in un anello di matrici
A = np.array([[1, 2], [3, 4]])
zero_matrice = np.zeros((2, 2), dtype=int)
print("A @ 0 =\n", A @ zero_matrice)
print("0 @ A =\n", zero_matrice @ A)

# Verifica della regola dei segni: (-a)*b = -(a*b)
a, b = 5, -3
print(f"\n(-a)*b = {(-a)*b}, -(a*b) = {-(a*b)}, uguali? {(-a)*b == -(a*b)}")

# Divisori dello zero in Z/6Z
n = 6
print(f"\nDivisori dello zero in Z/{n}Z:")
for a in range(1, n):
    for b in range(1, n):
        if (a * b) % n == 0:
            print(f"  {a} * {b} = {a*b} ≡ 0 (mod {n})")
````

## Esercizio svolto e commentato

**Problema:** determina per quali $n$ l'anello $\mathbb Z/n\mathbb Z$ **non** ha divisori dello zero, e dimostra la caratterizzazione trovata.

**Ragionamento:**

**Passo 1 — congetturiamo dall'esempio.** In $\mathbb Z/6\mathbb Z$ (con $6=2\times3$ composto) abbiamo trovato divisori dello zero ($2\times3\equiv0$). Congetturiamo che il problema sorga esattamente quando $n$ è **composto**, e che $\mathbb Z/p\mathbb Z$ con $p$ **primo** non abbia divisori dello zero.

**Passo 2 — dimostriamo che se $n$ è composto, $\mathbb Z/n\mathbb Z$ ha divisori dello zero.** Se $n=ab$ con $1<a,b<n$ (fattorizzazione non banale, possibile perché $n$ è composto), allora in $\mathbb Z/n\mathbb Z$: $[a]\times[b]=[ab]=[n]=[0]$, con $[a]\ne[0]$ e $[b]\ne[0]$ (perché $0<a,b<n$). Quindi $[a]$ e $[b]$ sono divisori dello zero.

**Passo 3 — dimostriamo che se $p$ è primo, $\mathbb Z/p\mathbb Z$ non ha divisori dello zero.** Supponiamo $[a]\times[b]=[0]$ in $\mathbb Z/p\mathbb Z$, cioè $p\mid ab$. Per una proprietà fondamentale dei numeri primi (se $p$ è primo e $p\mid ab$, allora $p\mid a$ oppure $p\mid b$ — dimostrabile con l'identità di Bézout, che esula da questo notebook ma è un fatto standard di teoria dei numeri elementare, coerente con l'uso già fatto di "3 primo" nell'esercizio 3 del notebook 10.6): se $p\mid a$, allora $[a]=[0]$; se $p\mid b$, allora $[b]=[0]$. In ogni caso, uno dei due fattori è $[0]$: non esistono divisori dello zero non nulli.

**Conclusione:** $\mathbb Z/n\mathbb Z$ non ha divisori dello zero $\iff$ $n$ è primo (o $n=1$, caso banale con un solo elemento). **Perché questo è importante**: $\mathbb Z/p\mathbb Z$ per $p$ primo è, come vedremo nel notebook 11.10, non solo privo di divisori dello zero ma addirittura un **campo** — ogni elemento non nullo ha inverso moltiplicativo — la struttura più ricca possibile per un anello finito, ed è la base per costruire i campi finiti usati in crittografia e teoria dei codici.

````python
def ha_divisori_zero(n):
    for a in range(1, n):
        for b in range(1, n):
            if (a * b) % n == 0:
                return True, (a, b)
    return False, None

for n in range(2, 13):
    trovato, coppia = ha_divisori_zero(n)
    stato = f"SI (esempio: {coppia[0]}*{coppia[1]})" if trovato else "NO"
    print(f"n={n}: divisori dello zero? {stato}")
````

**1.** È un anello: $(\mathbb Z\times\mathbb Z,+)$ è un gruppo abeliano (componente per componente), il prodotto è associativo e distributivo (ereditati da $\mathbb Z$ componente per componente), con neutro moltiplicativo $(1,1)$. **Ha divisori dello zero**: $(1,0)\times(0,1)=(0,0)$, con $(1,0)\ne(0,0)$ e $(0,1)\ne(0,0)$.

**2.** $(a+b)^2=(a+b)\times(a+b)=(a+b)\times a+(a+b)\times b$ (distributiva a destra) $=a\times a+b\times a+a\times b+b\times b$ (distributiva a sinistra su ciascun termine) $=a^2+ba+ab+b^2$. Usando la **commutatività** ($ba=ab$): $=a^2+ab+ab+b^2=a^2+2ab+b^2$. Senza commutatività (per esempio nell'anello delle matrici), si fermerebbe a $a^2+ba+ab+b^2$, che non si semplifica in $a^2+2ab+b^2$ a meno che $ab=ba$.

**3.** Supponiamo $1$ e $1'$ entrambi neutri moltiplicativi: $1=1\times1'$ (perché $1'$ è neutro) $=1'$ (perché $1$ è neutro): stesso argomento usato per l'unicità del neutro additivo nei gruppi (notebook 11.1), applicato qui alla struttura moltiplicativa.

**4.** Calcoliamo $(1-a)(1+a+a^2+\cdots+a^{k-1})$: sviluppando per distributiva, $=1+a+a^2+\cdots+a^{k-1}-a-a^2-\cdots-a^{k-1}-a^k$ (una somma telescopica, lo stesso meccanismo di cancellazione visto nel notebook 10.10) $=1-a^k=1-0=1$ (usando l'ipotesi $a^k=0$). Analogamente $(1+a+\cdots+a^{k-1})(1-a)=1-a^k=1$ (l'ordine non conta se assumiamo, come tipico in questo contesto, che $a$ commuti con se stesso — sempre vero). Dunque $1+a+\cdots+a^{k-1}$ soddisfa la definizione di inverso di $1-a$: $(1-a)^{-1}=1+a+\cdots+a^{k-1}$.

Verifica computazionale (esercizio 4, con matrici nilpotenti):

A = np.array([[0, 1], [0, 0]])  # nilpotente: A^2 = 0

I = np.eye(2, dtype=int)

## Riepilogo finale

- Un **anello** è un gruppo abeliano rispetto alla somma, con un prodotto associativo collegato dalla distributiva — senza richiedere inversi moltiplicativi.
- Dagli assiomi discendono, con dimostrazioni formali, fatti "ovvi" come $a\times0=0$ e la regola dei segni $(-a)b=-(ab)$: veri in **ogni** anello.
- I **divisori dello zero** (elementi non nulli il cui prodotto è zero) esistono in $\mathbb Z/n\mathbb Z$ esattamente quando $n$ è composto; $\mathbb Z/p\mathbb Z$ con $p$ primo ne è privo.
- Questa distinzione anticipa la gerarchia di strutture sempre più ricche che seguirà: anelli generali, domini integrali (senza divisori dello zero), campi (con inversi moltiplicativi).
