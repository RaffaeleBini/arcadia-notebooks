## Perché ci interessa

Nel notebook 11.6 abbiamo scoperto che $\mathbb Z/6\mathbb Z$ ha **divisori dello zero** ($2\times3=0$), mentre $\mathbb Z$ e $\mathbb Z/p\mathbb Z$ (per $p$ primo) non ne hanno. Questa distinzione non è un dettaglio tecnico: è **esattamente** ciò che permette, o impedisce, di "semplificare" un'equazione — un'operazione che facciamo costantemente con i numeri reali senza mai dubitarne, ma che può fallire clamorosamente in anelli più generali (per esempio con le matrici: $AB=AC$ non implica $B=C$ se $A$ non è invertibile). Questo notebook isola la classe di anelli — i **domini integrali** — dove questa cancellazione è sempre lecita, l'ultimo passo prima della definizione di campo (notebook 11.10).

## Teoria, passo dopo passo

### Definizione di dominio integrale

Un **dominio integrale** è un anello commutativo unitario (con $1\ne0$) **senza divisori dello zero**: $a\times b=0\Rightarrow a=0\text{ oppure }b=0$.

### La legge di cancellazione

**Teorema.** In un dominio integrale, $a\ne0$ e $ab=ac\Rightarrow b=c$.

**Dimostrazione.** $ab=ac\Rightarrow ab-ac=0\Rightarrow a(b-c)=0$ (distributiva, notebook 0.5, e la regola dei segni del notebook 11.6 per riscrivere $-ac=a(-c)$). Poiché $a\ne0$ e non ci sono divisori dello zero, deve essere $b-c=0$, cioè $b=c$. $\blacksquare$

Questa è la proprietà che rende un dominio integrale "sicuro" per i calcoli algebrici quotidiani: possiamo cancellare un fattore comune non nullo, esattamente come facciamo con i numeri interi o reali, ma **non** come potremmo fare, per esempio, con le matrici o con $\mathbb Z/6\mathbb Z$.

### Esempi e non-esempi

- $\mathbb Z$: dominio integrale (fatto elementare, usato implicitamente ovunque nel corso).
- $\mathbb Z/p\mathbb Z$ per $p$ **primo**: dominio integrale, dimostrato nell'esercizio svolto del notebook 11.6.
- $\mathbb Z/n\mathbb Z$ per $n$ **composto**: **non** è un dominio integrale (ha divisori dello zero, stesso esercizio).
- $\mathbb R[x]$ (polinomi reali): dominio integrale — se $p(x)q(x)=0$ come polinomio identicamente nullo, allora (confrontando i coefficienti del termine di grado massimo, che si moltiplicano senza possibilità di cancellazione essendo $\mathbb R$ un campo) uno dei due fattori deve essere il polinomio nullo.
- $M_n(\mathbb R)$ (matrici $n\times n$, $n\ge2$): **non** è un dominio integrale, e non è nemmeno commutativo — un doppio motivo di esclusione. Esempio esplicito: $\begin{pmatrix}1&0\\0&0\end{pmatrix}\begin{pmatrix}0&0\\0&1\end{pmatrix}=\begin{pmatrix}0&0\\0&0\end{pmatrix}$, con entrambi i fattori non nulli.
- $\mathbb Z\times\mathbb Z$: **non** è un dominio integrale (esercizio 1 del notebook 11.6: $(1,0)\times(0,1)=(0,0)$).

### Ogni campo è un dominio integrale

Anticipiamo un fatto che il notebook 11.10 formalizzerà: se ogni elemento non nullo ha un inverso moltiplicativo (proprietà di **campo**), allora automaticamente non ci sono divisori dello zero. **Dimostrazione**: se $ab=0$ con $a\ne0$, moltiplicando per $a^{-1}$ (che esiste per ipotesi di campo): $a^{-1}(ab)=a^{-1}\times0=0$, cioè $(a^{-1}a)b=0$, cioè $b=0$. Dunque ogni campo è un dominio integrale — ma **non viceversa**: $\mathbb Z$ è un dominio integrale ma non un campo (nessun intero eccetto $\pm1$ ha inverso moltiplicativo intero). I domini integrali sono quindi una classe **intermedia**, tra gli anelli generali e i campi.

### Domini a ideali principali (PID)

Un dominio integrale in cui **ogni** ideale è principale (generato da un solo elemento, notebook 11.7) si chiama **dominio a ideali principali** (PID, dall'inglese *principal ideal domain*). Il notebook 11.7 ha dimostrato che $\mathbb Z$ ha questa proprietà (ogni ideale $(a,b)=(\gcd(a,b))$); anche $\mathbb R[x]$ è un PID (per la divisione euclidea tra polinomi, che vedremo nel notebook 11.11), ma $\mathbb Z[x]$ **non** lo è: l'ideale $(2,x)=\{2p(x)+xq(x):p,q\in\mathbb Z[x]\}$ (i polinomi con termine noto pari) non è generato da un singolo polinomio — un fatto che segna una differenza strutturale sottile ma importante tra $\mathbb Z[x]$ e $\mathbb R[x]$, dovuta al fatto che $\mathbb Z$ non è un campo mentre $\mathbb R$ lo è.

````python
import numpy as np
import sympy as sp

# Verifica: le matrici NON formano un dominio integrale
A = np.array([[1, 0], [0, 0]])
B = np.array([[0, 0], [0, 1]])
print("A =\n", A)
print("B =\n", B)
print("A @ B =\n", A @ B)
print("A e B sono entrambe non nulle, ma il loro prodotto e' nullo: divisori dello zero.")

# Verifica: R[x] E' un dominio integrale (nessun esempio di p*q=0 con p,q non nulli)
x = sp.symbols('x')
p = x**2 + 1
q = x - 3
prodotto = sp.expand(p * q)
print(f"\n({p}) * ({q}) = {prodotto}")
print("Il prodotto e' il polinomio nullo solo se p=0 o q=0 (mai in questo caso, come atteso in un dominio integrale).")
````

## Esercizio svolto e commentato

**Problema:** dimostra che se $D$ è un dominio integrale **finito**, allora $D$ è un campo (ogni elemento non nullo ha inverso moltiplicativo) — un risultato che spiega perché $\mathbb Z/p\mathbb Z$ (finito, dominio integrale per $p$ primo) è automaticamente un campo, anticipando il notebook 11.10.

**Ragionamento:**

**Passo 1 — fissiamo un elemento non nullo.** Sia $a\in D$, $a\ne0$. Vogliamo trovare $a^{-1}\in D$.

**Passo 2 — consideriamo la funzione "moltiplicazione per $a$".** Definiamo $f_a:D\to D$, $f_a(x)=ax$. Dimostriamo che $f_a$ è **iniettiva**: se $f_a(x)=f_a(y)$, cioè $ax=ay$, per la legge di cancellazione (dimostrata nella teoria, valida perché $D$ è un dominio integrale e $a\ne0$) segue $x=y$.

**Passo 3 — usiamo la finitezza.** $f_a:D\to D$ è una funzione iniettiva **da un insieme finito in se stesso**. Per un fatto elementare di teoria degli insiemi finiti (una funzione iniettiva tra insiemi finiti della stessa cardinalità è automaticamente **suriettiva** — se non lo fosse, l'immagine avrebbe meno elementi del dominio, contraddicendo l'iniettività che richiede immagini distinte per punti distinti), $f_a$ è anche suriettiva.

**Passo 4 — troviamo l'inverso.** Per suriettività, esiste $b\in D$ con $f_a(b)=1$ (il neutro moltiplicativo esiste per ipotesi di dominio integrale unitario), cioè $ab=1$. Dunque $b=a^{-1}$.

**Conclusione:** ogni elemento non nullo di $D$ ha inverso moltiplicativo: $D$ è un campo. **Perché la finitezza è essenziale**: in un dominio integrale **infinito** come $\mathbb Z$, la stessa funzione $f_2(x)=2x$ è iniettiva ma **non** suriettiva (nessun $x\in\mathbb Z$ soddisfa $2x=1$) — l'argomento del passo 3 fallisce esplicitamente per insiemi infiniti (coerente con la distinzione tra funzioni iniettive e suriettive su insiemi infiniti già osservata nel notebook 10.5, dove una funzione può essere iniettiva senza essere suriettiva proprio a causa dell'infinità del dominio), ed è esattamente per questo che $\mathbb Z$ resta un dominio integrale senza essere un campo.

````python
def moltiplicazione_per_a(a, n):
    """Restituisce la funzione x -> a*x mod n, come dizionario."""
    return {x: (a * x) % n for x in range(n)}

# Verifica su Z/7Z (dominio integrale finito, quindi campo, essendo 7 primo)
n = 7
for a in range(1, n):
    f_a = moltiplicazione_per_a(a, n)
    immagine = sorted(f_a.values())
    iniettiva = len(set(immagine)) == n
    suriettiva = immagine == list(range(n))
    inverso = [x for x, y in f_a.items() if y == 1][0]
    print(f"a={a}: iniettiva={iniettiva}, suriettiva={suriettiva}, inverso di {a} e' {inverso} "
          f"(verifica: {a}*{inverso} mod {n} = {(a*inverso)%n})")
````

## Riepilogo finale

- Un **dominio integrale** è un anello commutativo unitario senza divisori dello zero: garantisce la **legge di cancellazione** $ab=ac,a\ne0\Rightarrow b=c$.
- $\mathbb Z$, $\mathbb Z/p\mathbb Z$ ($p$ primo), $\mathbb R[x]$ sono domini integrali; $\mathbb Z/n\mathbb Z$ ($n$ composto), le matrici, $\mathbb Z\times\mathbb Z$ non lo sono.
- **Ogni campo è un dominio integrale**, ma non viceversa ($\mathbb Z$ è un controesempio); tuttavia **ogni dominio integrale finito è un campo** — un risultato che chiude il cerchio per gli anelli finiti.
- Un **dominio a ideali principali** (PID) è un dominio integrale in cui ogni ideale è principale: $\mathbb Z$ e $\mathbb R[x]$ lo sono, $\mathbb Z[x]$ no.
