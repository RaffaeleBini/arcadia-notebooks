# 11.7 — Ideali

**Prerequisiti:** 11.6 Anelli, 11.2 Sottogruppi, 11.3 Gruppi quoziente.


## Perché ci interessa

Nel notebook 11.3 abbiamo scoperto che non ogni sottogruppo permette di costruire un gruppo quoziente ben definito: serve la **normalità**. Per gli anelli, che hanno due operazioni, la situazione è analoga ma più delicata: un semplice "sottoanello" (chiuso rispetto a somma e prodotto) non basta a costruire un anello quoziente. Serve una nozione più forte — l'**ideale** — che gioca, per gli anelli, lo stesso ruolo che i sottogruppi normali giocano per i gruppi. Il prototipo è $n\mathbb Z\subset\mathbb Z$: già usato come sottogruppo nel notebook 11.2, ora ne riveliamo la proprietà aggiuntiva che lo rende un ideale.

## Teoria, passo dopo passo

### Sottoanelli: non bastano

Un **sottoanello** $S\subseteq R$ è un sottoinsieme che, con le stesse operazioni, è esso stesso un anello: chiuso rispetto a somma, opposto, e prodotto. Per esempio, $\mathbb Z\subset\mathbb Q$ è un sottoanello. Ma se proviamo a costruire $\mathbb Q/\mathbb Z$ definendo il prodotto tra classi $[a]\times[b]=[ab]$, la definizione **non è ben posta**: per esempio $[1/2]=[3/2]$ in $\mathbb Q/\mathbb Z$ (differiscono per l'intero $1$), ma moltiplicando entrambi per $[1/2]$: $[1/2]\times[1/2]=[1/4]$, mentre $[3/2]\times[1/2]=[3/4]$, e $[1/4]\ne[3/4]$ (differiscono per $1/2\notin\mathbb Z$) — il prodotto dipende dal rappresentante scelto. Serve una condizione più forte di "sottoanello" per evitare questo problema.

### Definizione di ideale

Un sottoinsieme $I\subseteq R$ (con $R$ anello commutativo, per semplicità) è un **ideale** se:

1. $(I,+)$ è un sottogruppo di $(R,+)$ (in particolare $0\in I$, chiuso per somma e opposto);
2. **Assorbimento**: per ogni $r\in R$ e $a\in I$, $r\times a\in I$ (non solo per $r\in I$: per **qualunque** $r$ dell'intero anello).

La condizione 2 è molto più forte della semplice chiusura di un sottoanello (che richiederebbe solo $a\times b\in I$ per $a,b\in I$): un ideale "assorbe" la moltiplicazione per **qualsiasi** elemento dell'anello, non solo per i propri elementi. È esattamente questa forza aggiuntiva a garantire (come nel caso dei sottogruppi normali) che il quoziente sia ben definito.

### $n\mathbb Z$ è un ideale di $\mathbb Z$

Verifichiamo: $(n\mathbb Z,+)$ è un sottogruppo di $(\mathbb Z,+)$ (notebook 11.2). Per l'assorbimento: dato $r\in\mathbb Z$ qualsiasi e $a=nk\in n\mathbb Z$, $r\times a=r\times nk=n(rk)\in n\mathbb Z$ (proprietà associativa e commutativa di $\mathbb Z$, notebook 0.5): l'assorbimento vale. Dunque $n\mathbb Z\trianglelefteq\mathbb Z$ è un ideale — non solo un sottogruppo, come nel notebook 11.2, ma un ideale nel senso pieno.

### Ideali principali

Dato $a\in R$, l'**ideale principale generato da $a$** è $(a)=\{r\times a:r\in R\}$ (in un anello commutativo unitario) — il più piccolo ideale che contiene $a$. Si verifica facilmente con la definizione: $(a)$ è chiuso per somma ($ra+sa=(r+s)a$, distributiva), per opposto ($-(ra)=(-r)a$), e per assorbimento ($s\times(ra)=(sr)a\in(a)$, associatività e commutatività). In $\mathbb Z$, ogni ideale è principale: $n\mathbb Z=(n)$ per ogni $n$ — un fatto che riflette la struttura molto ordinata di $\mathbb Z$ (un **dominio a ideali principali**, terminologia che approfondiremo nel notebook 11.9).

### Somma e intersezione di ideali

**Teorema.** Se $I,J$ sono ideali di $R$, allora $I\cap J$ e $I+J=\{a+b:a\in I,b\in J\}$ sono ideali.

**Dimostrazione (intersezione).** $I\cap J\ne\emptyset$ (contiene $0$). È un sottogruppo additivo (intersezione di sottogruppi, esercizio 3 del notebook 11.2). Assorbimento: se $x\in I\cap J$ e $r\in R$, allora $rx\in I$ (assorbimento in $I$) e $rx\in J$ (assorbimento in $J$), quindi $rx\in I\cap J$.

**Dimostrazione (somma).** $I+J\ne\emptyset$ (contiene $0+0=0$). Sottogruppo additivo: $(a_1+b_1)-(a_2+b_2)=(a_1-a_2)+(b_1-b_2)\in I+J$ (usando che $I,J$ sono sottogruppi). Assorbimento: $r(a+b)=ra+rb\in I+J$ (distributiva, e $ra\in I$, $rb\in J$ per assorbimento in ciascuno). $\blacksquare$

Questi ideali "combinati" saranno gli strumenti con cui, nel notebook 11.11, descriveremo la struttura degli ideali nell'anello dei polinomi.

````python
from sympy import gcd, ideal, symbols, QQ
from sympy.abc import x

# Verifica che nZ e' un ideale di Z: chiusura per somma e assorbimento
n = 4
elementi_ideale = [n*k for k in range(-5, 6)]
print(f"Elementi di {n}Z (campione):", elementi_ideale)

# Assorbimento: r * a per r arbitrario, a in nZ, resta in nZ
import random
random.seed(0)
for _ in range(5):
    r = random.randint(-20, 20)
    a = random.choice(elementi_ideale)
    prodotto = r * a
    print(f"r={r}, a={a}: r*a={prodotto}, appartiene a {n}Z? {prodotto % n == 0}")

# Ideale generato da un polinomio in Q[x]: (x^2 - 1)
I = ideal(x**2 - 1, QQ.old_poly_ring(x))
print("\nIdeale generato da x^2-1 in Q[x]:", I)
````

## Esercizio svolto e commentato

**Problema:** dimostra che in $\mathbb Z$, l'ideale generato da due interi $a,b$, cioè $(a,b)=\{ra+sb:r,s\in\mathbb Z\}$ (la somma degli ideali principali $(a)+(b)$), coincide con l'ideale principale $(d)$ dove $d=\gcd(a,b)$.

**Ragionamento:**

**Passo 1 — dimostriamo $(d)\subseteq(a,b)$.** Per l'identità di Bézout (un fatto di teoria dei numeri elementare: il massimo comun divisore di $a,b$ si scrive sempre come combinazione intera $d=ra+sb$ per opportuni $r,s\in\mathbb Z$), $d\in(a,b)$. Poiché $(a,b)$ è un ideale (somma di ideali, dimostrata nella teoria) e $d\in(a,b)$, ogni multiplo di $d$ è in $(a,b)$ per assorbimento: $(d)\subseteq(a,b)$.

**Passo 2 — dimostriamo $(a,b)\subseteq(d)$.** Ogni elemento di $(a,b)$ è della forma $ra+sb$. Poiché $d\mid a$ e $d\mid b$ (per definizione di massimo comun divisore), scriviamo $a=da'$, $b=db'$: allora $ra+sb=r(da')+s(db')=d(ra'+sb')$, che è un multiplo di $d$, quindi appartiene a $(d)$.

**Conclusione:** $(a,b)=(d)$ con $d=\gcd(a,b)$. **Perché questo risultato è potente**: mostra che in $\mathbb Z$ ogni ideale generato da più elementi collassa sempre a un **ideale principale** generato dal loro MCD — coerente con l'osservazione della teoria che ogni ideale di $\mathbb Z$ è principale. Questa proprietà, tutt'altro che scontata in un anello generico (esistono anelli con ideali non principali, per esempio $(2,x)$ in $\mathbb Z[x]$), è ciò che rende $\mathbb Z$ un **dominio a ideali principali**, terminologia che ritroveremo nel notebook 11.9.

````python
import sympy as sp

a, b = 12, 18
d = sp.gcd(a, b)
print(f"gcd({a},{b}) = {d}")

# Identita' di Bezout: troviamo r,s tali che r*a + s*b = d
r, s, gcd_val = sp.gcdex(a, b)
print(f"Identita' di Bezout: r={r}, s={s}, r*a+s*b = {r*a + s*b} (atteso = {d})")

# Verifica: ogni combinazione r*a+s*b e' multiplo di d
import random
random.seed(1)
for _ in range(5):
    r_test = random.randint(-10, 10)
    s_test = random.randint(-10, 10)
    combinazione = r_test*a + s_test*b
    print(f"r={r_test}, s={s_test}: {r_test}*{a}+{s_test}*{b} = {combinazione}, "
          f"multiplo di {d}? {combinazione % d == 0}")
````

## Riepilogo finale

- Un **ideale** $I\subseteq R$ è un sottogruppo additivo che **assorbe** la moltiplicazione per ogni elemento dell'anello: $r\in R,a\in I\Rightarrow ra\in I$ — condizione più forte del semplice sottoanello.
- Gli **ideali principali** $(a)=\{ra:r\in R\}$ sono i più semplici; in $\mathbb Z$, ogni ideale è principale, e $(a,b)=(\gcd(a,b))$ (identità di Bézout).
- Somma e intersezione di ideali sono sempre ideali.
- Gli ideali sono, per gli anelli, l'esatto analogo dei sottogruppi normali per i gruppi: il prossimo passo naturale è costruire, su di essi, un **anello quoziente**.
