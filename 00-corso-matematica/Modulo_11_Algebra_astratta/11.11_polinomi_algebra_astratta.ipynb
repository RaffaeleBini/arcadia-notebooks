## Teoria, passo dopo passo

### L'anello $F[x]$ e il grado

Per un campo $F$, $F[x]$ è l'anello dei polinomi $a_0+a_1x+\cdots+a_nx^n$ con $a_i\in F$ (notebook 1.1, ora a coefficienti in un campo arbitrario, non solo $\mathbb R$). Il **grado** $\deg p$ è l'indice del coefficiente non nullo più alto. Proprietà fondamentale: $\deg(pq)=\deg p+\deg q$ (i termini di grado massimo si moltiplicano senza cancellarsi, essendo $F$ un campo — quindi senza divisori dello zero, notebook 11.9). Questa proprietà è precisamente il motivo per cui $F[x]$ è un dominio integrale (già usato nel notebook 11.9): se $pq=0$ con $p,q\ne0$, si avrebbe $\deg(pq)=\deg p+\deg q\ge0$, ma il polinomio nullo non ha un grado ben definito in questo senso — contraddizione formale che conferma l'assenza di divisori dello zero.

### Divisione euclidea tra polinomi

**Teorema (divisione con resto).** Dati $p(x),d(x)\in F[x]$ con $d\ne0$, esistono unici $q(x),r(x)\in F[x]$ tali che $p(x)=q(x)d(x)+r(x)$, con $r=0$ oppure $\deg r<\deg d$.

**Dimostrazione (esistenza, per induzione forte sul grado di $p$, notebook 10.4).** *Base*: se $\deg p<\deg d$ (incluso $p=0$), basta $q=0,r=p$. *Passo induttivo*: supponiamo $\deg p\ge\deg d$, e che il teorema valga per ogni polinomio di grado minore di $\deg p$. Siano $a_n$ il coefficiente direttore di $p$ (grado $n$) e $b_m$ quello di $d$ (grado $m\le n$). Poiché $F$ è un campo, $b_m^{-1}$ esiste (proprietà cruciale di campo, notebook 11.10 — è qui che serve $F$ campo e non un anello generico!). Formiamo $p_1(x)=p(x)-\dfrac{a_n}{b_m}x^{n-m}d(x)$: il termine di grado $n$ si cancella per costruzione, quindi $\deg p_1<\deg p$ (o $p_1=0$). Per ipotesi induttiva, $p_1=q_1d+r$ con $r=0$ o $\deg r<\deg d$. Sostituendo: $p=p_1+\dfrac{a_n}{b_m}x^{n-m}d=\left(q_1+\dfrac{a_n}{b_m}x^{n-m}\right)d+r$: il quoziente cercato è $q=q_1+\dfrac{a_n}{b_m}x^{n-m}$.

**Unicità**: se $p=q_1d+r_1=q_2d+r_2$ con entrambi i resti di grado $<\deg d$ (o nulli), allora $(q_1-q_2)d=r_2-r_1$; se $q_1\ne q_2$, il membro sinistro ha grado $\ge\deg d$ (proprietà del grado del prodotto), ma il destro ha grado $<\deg d$: contraddizione, quindi $q_1=q_2$ e di conseguenza $r_1=r_2$. $\blacksquare$

**Il ruolo essenziale di $F$ campo**: il passo cruciale usa $b_m^{-1}$ — se $F$ fosse solo un anello (per esempio $\mathbb Z$), la divisione euclidea tra polinomi potrebbe fallire (in $\mathbb Z[x]$, dividere $x^2$ per $2x$ non dà un quoziente intero). È esattamente questo il motivo per cui $\mathbb Z[x]$ non è un PID (notebook 11.9), mentre $F[x]$ per $F$ campo lo è sempre.

### Polinomi irriducibili

Un polinomio $p(x)\in F[x]$ non costante è **irriducibile** se non si scompone come prodotto di due polinomi di grado inferiore (entrambi $\ge1$) — l'analogo esatto dei numeri primi in $\mathbb Z$. Per esempio, $x^2+1$ è irriducibile in $\mathbb R[x]$ (non ha radici reali, quindi non si fattorizza in due lineari), ma **si fattorizza** in $\mathbb C[x]$: $x^2+1=(x-i)(x+i)$ — l'irriducibilità **dipende dal campo** su cui si lavora, una differenza fondamentale rispetto ai numeri primi (che sono primi indipendentemente dal contesto).

### Fattorizzazione unica

**Teorema (enunciato, senza dimostrazione completa).** Ogni polinomio non costante di $F[x]$ si scrive in modo unico (a meno di riordino e di fattori costanti) come prodotto di polinomi irriducibili — esattamente come ogni intero si fattorizza in modo unico in numeri primi (teorema fondamentale dell'aritmetica). La dimostrazione ricalca da vicino quella per $\mathbb Z$: si basa sulla divisione euclidea (che fornisce un algoritmo di Euclide per il MCD tra polinomi) e sull'identità di Bézout per polinomi, in perfetta analogia con l'esercizio svolto del notebook 11.7.

````python
import sympy as sp

x = sp.symbols('x')

# Divisione euclidea tra polinomi
p = x**4 + 2*x**3 - x + 5
d = x**2 - 1

q, r = sp.div(p, d, x)
print(f"p(x) = {p}")
print(f"d(x) = {d}")
print(f"quoziente q(x) = {q}")
print(f"resto r(x) = {r}")
print(f"Verifica: q*d + r = {sp.expand(q*d + r)}, uguale a p? {sp.expand(q*d+r-p) == 0}")

# Irriducibilita' dipende dal campo: x^2+1 su R vs su C
print("\nFattorizzazione di x^2+1 su Q/R:", sp.factor(x**2 + 1))
print("Fattorizzazione di x^2+1 su C:", sp.factor(x**2 + 1, extension=[sp.I]))
````

## Esercizio svolto e commentato

**Problema:** usa l'algoritmo di Euclide (analogo a quello per interi, esercizio svolto del notebook 11.7) per calcolare $\gcd(p,q)$ tra $p(x)=x^3-1$ e $q(x)=x^2-1$ in $\mathbb Q[x]$, e verifica il risultato tramite le radici.

**Ragionamento:**

**Passo 1 — dividiamo $p$ per $q$.** $x^3-1=(x)(x^2-1)+(x-1)$ (verifica: $x\cdot(x^2-1)=x^3-x$, e $x^3-1-(x^3-x)=x-1$): quoziente $x$, resto $x-1$.

**Passo 2 — applichiamo l'algoritmo di Euclide, dividendo $q$ per il resto.** $x^2-1=(x+1)(x-1)+0$ (verifica: $(x+1)(x-1)=x^2-1$): resto $0$.

**Passo 3 — l'ultimo resto non nullo è il MCD.** Il resto della divisione precedente era $x-1$, e ora il resto è $0$: per l'algoritmo di Euclide (stesso principio usato per gli interi), $\gcd(p,q)=x-1$ (a meno di un fattore costante, per convenzione si normalizza a coefficiente direttore $1$).

**Passo 4 — verifichiamo tramite le radici.** $p(x)=x^3-1=(x-1)(x^2+x+1)$ (differenza di cubi, notebook 1.2 generalizzata) ha come unica radice reale $x=1$. $q(x)=x^2-1=(x-1)(x+1)$ ha radici $x=1,-1$. L'unico fattore lineare **comune** alle due fattorizzazioni è $(x-1)$: coerente con $\gcd(p,q)=x-1$.

**Conclusione:** $\gcd(x^3-1,x^2-1)=x-1$ in $\mathbb Q[x]$. **L'analogia con $\mathbb Z$ è completa**: l'algoritmo di Euclide per polinomi funziona esattamente come per interi (notebook 11.7), con la divisione euclidea di polinomi al posto della divisione con resto tra interi, e il grado al posto del valore assoluto come "misura di grandezza" che decresce a ogni passo, garantendo la terminazione dell'algoritmo in un numero finito di passi.

````python
import sympy as sp

x = sp.symbols('x')
p = x**3 - 1
q = x**2 - 1

mcd = sp.gcd(p, q, x)
print(f"gcd({p}, {q}) = {mcd}")

# Verifica tramite fattorizzazione
print("\nFattorizzazione di p:", sp.factor(p))
print("Fattorizzazione di q:", sp.factor(q))

# Algoritmo di Euclide passo-passo
a, b = p, q
while b != 0:
    quoziente, resto = sp.div(a, b, x)
    print(f"{a} = ({quoziente})*({b}) + ({resto})")
    a, b = b, resto
print(f"\nUltimo resto non nullo (MCD, a meno di costante): {a}")
````

## Riepilogo finale

- $F[x]$, per $F$ campo, ha una **divisione euclidea** con resto (come $\mathbb Z$), possibile grazie all'invertibilità del coefficiente direttore — un punto dove la struttura di campo è essenziale.
- I **polinomi irriducibili** sono l'analogo dei numeri primi; l'irriducibilità **dipende dal campo**: $x^2+1$ è irriducibile su $\mathbb R$ ma non su $\mathbb C$.
- $F[x]$ ha **fattorizzazione unica** in irriducibili, e l'**algoritmo di Euclide** calcola il MCD tra polinomi esattamente come tra interi.
- Questa struttura è ciò che rende $F[x]$ un dominio a ideali principali (notebook 11.9), e sarà lo strumento per costruire estensioni di campo come $F[x]/(p(x))$ nel prossimo notebook, generalizzando $\mathbb C=\mathbb R[x]/(x^2+1)$.
