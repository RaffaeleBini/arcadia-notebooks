## Perché ci interessa

Nel notebook 11.1 abbiamo usato le rotazioni di un triangolo come primo esempio di gruppo, rappresentandole come permutazioni dei vertici. Questo notebook rende quell'idea centrale: il gruppo $S_n$ di **tutte** le permutazioni di $n$ oggetti è, in un certo senso, il gruppo "universale" — il **teorema di Cayley** dimostrerà che ogni gruppo finito (per quanto astratto) è isomorfo a un sottogruppo di qualche $S_n$. Studiamo $S_n$ con gli strumenti giusti — la notazione ciclica — e la distinzione tra permutazioni pari e dispari, che genera il sottogruppo alternante $A_n$, protagonista della teoria di Galois nel notebook 11.13.

## Teoria, passo dopo passo

### Il gruppo simmetrico $S_n$

Una **permutazione** di $\{1,\ldots,n\}$ è una funzione biiettiva $\sigma:\{1,\ldots,n\}\to\{1,\ldots,n\}$ (notebook 3.5). L'insieme di tutte le permutazioni, con la composizione (notebook 3.7), forma un gruppo $S_n$: la composizione di biiezioni è una biiezione (chiusura), la composizione è associativa (proprietà generale della composizione di funzioni), l'identità è il neutro, e ogni biiezione ha un'inversa (notebook 3.6). $|S_n|=n!$ (il numero di modi di riordinare $n$ oggetti, un fatto dimostrabile per induzione, notebook 10.4: scegliere l'immagine di $1$ in $n$ modi, poi l'immagine di $2$ tra i restanti $n-1$, e così via). $S_n$ è **non abeliano** per $n\ge3$.

### Notazione ciclica

Un **ciclo** $(a_1\,a_2\,\ldots\,a_k)$ è la permutazione che manda $a_1\to a_2\to\cdots\to a_k\to a_1$, lasciando fissi gli altri elementi. Ogni permutazione si scrive come **prodotto di cicli disgiunti** (che commutano tra loro, agendo su elementi diversi): per esempio, in $S_5$, la permutazione che manda $1\to3,2\to2,3\to5,4\to1,5\to4$ si scrive $(1\,3\,5\,4)(2)$, o semplicemente $(1\,3\,5\,4)$ omettendo i punti fissi. Un ciclo di lunghezza $2$, $(a\,b)$, si chiama **trasposizione**.

### Ogni permutazione è un prodotto di trasposizioni

**Teorema.** Ogni ciclo $(a_1\,a_2\,\ldots\,a_k)$ si scrive come prodotto di $k-1$ trasposizioni: $(a_1\,a_2\,\ldots\,a_k)=(a_1\,a_k)(a_1\,a_{k-1})\cdots(a_1\,a_2)$.

**Dimostrazione (verifica diretta dell'azione).** Applichiamo il membro destro a $a_1$ (leggendo da destra a sinistra, come per la composizione di funzioni, notebook 3.7): $(a_1\,a_2)$ manda $a_1\to a_2$; le trasposizioni successive $(a_1\,a_{j})$ per $j\ge3$ non toccano $a_2$ (agiscono solo su $a_1$ e $a_j$), quindi il risultato finale è $a_2$: coerente con $(a_1\ldots a_k)$ che manda $a_1\to a_2$. Applichiamo il membro destro a $a_i$ per $2\le i\le k-1$: le trasposizioni $(a_1\,a_2),\ldots,(a_1\,a_{i-1})$ non toccano $a_i$; $(a_1\,a_i)$ manda $a_i\to a_1$; le successive $(a_1\,a_{i+1}),\ldots$ mandano $a_1\to a_{i+1}$ (essendo $a_1$ il risultato del passo precedente) — il risultato netto è $a_i\to a_{i+1}$, coerente. Un controllo analogo per $a_k$ (che deve tornare a $a_1$) completa la verifica per ogni elemento del ciclo, e i punti fissi restano ovviamente fissi. $\blacksquare$ Poiché ogni permutazione è prodotto di cicli disgiunti, e ogni ciclo è prodotto di trasposizioni: **ogni permutazione è un prodotto di trasposizioni**.

### Permutazioni pari e dispari: il gruppo alternante

Si dimostra (con un argomento che coinvolge il segno del prodotto $\prod_{i<j}(x_i-x_j)$ sotto l'azione della permutazione sugli indici, che esula dai dettagli qui) che, sebbene la scomposizione in trasposizioni **non sia unica**, la **parità** del numero di trasposizioni usate è sempre la stessa per una data permutazione. Questo permette di definire il **segno**: $\operatorname{sgn}(\sigma)=+1$ se $\sigma$ è prodotto di un numero **pari** di trasposizioni ("permutazione pari"), $-1$ se dispari.

**Teorema.** $\operatorname{sgn}:S_n\to(\{+1,-1\},\times)$ è un omomorfismo.

**Dimostrazione.** Se $\sigma$ è prodotto di $p$ trasposizioni e $\tau$ di $q$ trasposizioni, allora $\sigma\tau$ è prodotto di $p+q$ trasposizioni (semplicemente concatenando le due scomposizioni), quindi $\operatorname{sgn}(\sigma\tau)=(-1)^{p+q}=(-1)^p(-1)^q=\operatorname{sgn}(\sigma)\operatorname{sgn}(\tau)$. $\blacksquare$

Il **gruppo alternante** $A_n=\ker(\operatorname{sgn})$ è, per il notebook 11.4, un sottogruppo **normale** di $S_n$ (nucleo di un omomorfismo), di indice 2 (essendo $\operatorname{sgn}$ suriettiva su $\{\pm1\}$ per $n\ge2$, tramite una singola trasposizione), quindi $|A_n|=n!/2$. $A_n$ raccoglie tutte le permutazioni pari, ed è il protagonista, insieme a $S_5$, del notebook 11.13: la sua struttura (semplice, cioè priva di sottogruppi normali non banali, per $n\ge5$) è precisamente il motivo algebrico per cui l'equazione di quinto grado non ha una formula risolutiva generale.

### Il teorema di Cayley

**Teorema (Cayley).** Ogni gruppo $G$ è isomorfo a un sottogruppo di $S_G$ (il gruppo delle permutazioni dell'insieme $G$ stesso); se $|G|=n$ è finito, $G$ è isomorfo a un sottogruppo di $S_n$.

**Dimostrazione.** Per ogni $a\in G$, definiamo $\lambda_a:G\to G$, $\lambda_a(x)=a*x$ (**traslazione sinistra**). $\lambda_a$ è biiettiva: iniettiva per la legge di cancellazione (notebook 11.1), suriettiva perché per ogni $y\in G$, $x=a^{-1}*y$ soddisfa $\lambda_a(x)=y$. Dunque $\lambda_a\in S_G$. Definiamo $\Phi:G\to S_G$, $\Phi(a)=\lambda_a$. Verifichiamo che è un omomorfismo: $\lambda_{a*b}(x)=(a*b)*x=a*(b*x)=\lambda_a(\lambda_b(x))=(\lambda_a\circ\lambda_b)(x)$ per ogni $x$, quindi $\Phi(a*b)=\lambda_{a*b}=\lambda_a\circ\lambda_b=\Phi(a)\circ\Phi(b)$. $\Phi$ è iniettiva: se $\Phi(a)=\Phi(e)=\text{id}$, allora $\lambda_a(e)=a*e=a$ deve uguagliare $\lambda_e(e)=e$, quindi $a=e$ — nucleo banale, iniettiva per il criterio del notebook 11.4. Dunque $\Phi:G\to\operatorname{Im}\Phi\subseteq S_G$ è un isomorfismo su un sottogruppo di $S_G$. $\blacksquare$

**Il significato profondo**: nessun gruppo, per quanto costruito in modo astratto, sfugge a questa rappresentazione concreta come insieme di permutazioni — le permutazioni sono, in questo senso preciso, la struttura di gruppo "più generale possibile".

````python
from sympy.combinatorics import Permutation

# Notazione ciclica: una permutazione di S5
sigma = Permutation([2, 1, 4, 0, 3])  # manda 0->2, 1->1, 2->4, 3->0, 4->3 (indici da 0)
print("Permutazione (forma array, indici da 0):", sigma.array_form)
print("Notazione ciclica:", sigma.cyclic_form)

# Verifica del segno tramite scomposizione in trasposizioni
print("\nNumero minimo di trasposizioni (parita'):", sigma.length())
print("La permutazione e' pari?", sigma.is_even)
print("Segno:", 1 if sigma.is_even else -1)

# Verifica che sgn e' un omomorfismo: sgn(sigma*tau) = sgn(sigma)*sgn(tau)
tau = Permutation([1, 0, 2, 4, 3])
segno_sigma = 1 if sigma.is_even else -1
segno_tau = 1 if tau.is_even else -1
composta = sigma * tau
segno_composta = 1 if composta.is_even else -1

print(f"\nsgn(sigma)={segno_sigma}, sgn(tau)={segno_tau}, prodotto={segno_sigma*segno_tau}")
print(f"sgn(sigma*tau)={segno_composta}")
print("Coincidono (omomorfismo verificato)?", segno_composta == segno_sigma * segno_tau)
````

## Esercizio svolto e commentato

**Problema:** applica il teorema di Cayley esplicitamente a $\mathbb Z/3\mathbb Z$, trovando il sottogruppo di $S_3$ a cui è isomorfo.

**Ragionamento:**

**Passo 1 — identifichiamo gli elementi.** $\mathbb Z/3\mathbb Z=\{0,1,2\}$, con l'operazione di somma modulo 3. Etichettiamo gli elementi come posizioni $1,2,3$ (per usare la notazione di $S_3$, con $0\leftrightarrow1,1\leftrightarrow2,2\leftrightarrow3$).

**Passo 2 — calcoliamo le traslazioni sinistre $\lambda_a$.** $\lambda_0(x)=0+x=x$: l'identità, $()$. $\lambda_1(x)=1+x$: manda $0\to1,1\to2,2\to0$, cioè (nelle posizioni $1,2,3$) $1\to2,2\to3,3\to1$: il ciclo $(1\,2\,3)$. $\lambda_2(x)=2+x$: manda $0\to2,1\to0,2\to1$, cioè $1\to3,2\to1,3\to2$: il ciclo $(1\,3\,2)$.

**Passo 3 — verifichiamo che $\{\text{id},(1\,2\,3),(1\,3\,2)\}$ è un sottogruppo di $S_3$.** Componendo $(1\,2\,3)$ con se stesso: $(1\,2\,3)(1\,2\,3)$ manda $1\to2\to3$, cioè $1\to3$; $2\to3\to1$, cioè $2\to1$; $3\to1\to2$, cioè $3\to2$: risulta $(1\,3\,2)$, ancora nell'insieme. Chiusura verificata (e per un gruppo di 3 elementi generato da un solo ciclo, la struttura è automaticamente quella di $\mathbb Z/3\mathbb Z$, per il risultato dell'esercizio svolto del notebook 11.2 sui gruppi di ordine primo).

**Conclusione:** $\Phi:\mathbb Z/3\mathbb Z\to S_3$, $\Phi(0)=\text{id}$, $\Phi(1)=(1\,2\,3)$, $\Phi(2)=(1\,3\,2)$, è un isomorfismo tra $\mathbb Z/3\mathbb Z$ e il sottogruppo ciclico $\{\text{id},(1\,2\,3),(1\,3\,2)\}\le S_3$ — che è, non a caso, esattamente il sottogruppo delle rotazioni del gruppo diedrale $D_3$ incontrato nel notebook 11.2 (rotazioni del triangolo, viste come permutazioni dei vertici).

````python
from sympy.combinatorics import Permutation, PermutationGroup

# Costruiamo esplicitamente le traslazioni sinistre di Z/3Z come permutazioni di {0,1,2}
def lambda_a(a, n):
    return Permutation([(a + x) % n for x in range(n)])

n = 3
traslazioni = [lambda_a(a, n) for a in range(n)]
for a, perm in enumerate(traslazioni):
    print(f"lambda_{a}: {perm.array_form}, ciclica: {perm.cyclic_form}")

gruppo_immagine = PermutationGroup(traslazioni)
print("\nOrdine del gruppo immagine:", gruppo_immagine.order())
print("Coincide con |Z/3Z| = 3?", gruppo_immagine.order() == n)
````

## Riepilogo finale

- $S_n$, il gruppo di tutte le permutazioni di $n$ oggetti, ha ordine $n!$; ogni permutazione si scrive come prodotto di cicli disgiunti, e ogni ciclo come prodotto di trasposizioni.
- Il **segno** $\operatorname{sgn}:S_n\to\{\pm1\}$ è un omomorfismo; il suo nucleo, il **gruppo alternante** $A_n$, è normale e ha indice 2.
- Il **teorema di Cayley**: ogni gruppo è isomorfo a un sottogruppo di un gruppo di permutazioni, tramite le traslazioni sinistre — un risultato di rappresentazione universale.
- $A_n$ per $n\ge5$ sarà, nel notebook 11.13, l'ingrediente algebrico chiave per spiegare perché l'equazione generale di quinto grado non ha soluzione per radicali.
