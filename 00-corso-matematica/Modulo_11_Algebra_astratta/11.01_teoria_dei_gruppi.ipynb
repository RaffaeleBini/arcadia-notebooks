## Perché ci interessa

Nel notebook 0.5 abbiamo nominato per la prima volta le proprietà commutativa, associativa, l'elemento neutro e l'opposto — proprietà che da allora abbiamo usato, senza più pensarci, per i numeri, per i vettori (notebook 6.1), per le matrici (notebook 6.10), per i numeri complessi (notebook 9.1). Ma queste operazioni, così diverse tra loro (sommare vettori, moltiplicare matrici, comporre rotazioni), condividono una **struttura algebrica comune**. La **teoria dei gruppi** isola esattamente questa struttura, dimenticando *cosa* sono gli oggetti e *cosa* significa l'operazione, per studiare solo *come si comportano*. Il guadagno è enorme: un teorema dimostrato una volta sui gruppi vale automaticamente per ogni struttura che soddisfi gli assiomi — numeri, matrici, simmetrie geometriche, permutazioni — aprendo la porta all'algebra astratta, il linguaggio comune a gran parte della matematica moderna.

## Teoria, passo dopo passo

### Dal concreto all'astratto: cosa hanno in comune $(\mathbb Z,+)$ e le rotazioni di un triangolo?

Consideriamo $(\mathbb Z,+)$: la somma di interi è associativa, ha un elemento neutro ($0$), e ogni elemento ha un opposto. Consideriamo ora le **rotazioni di un triangolo equilatero** che lo sovrappongono a se stesso: rotazioni di $0°,120°,240°$. Componendo due rotazioni (prima una, poi l'altra) si ottiene ancora una rotazione dell'insieme; la composizione è associativa (come ogni composizione di funzioni, notebook 3.7); c'è una rotazione "neutra" ($0°$, che non cambia nulla); e ogni rotazione ha un'inversa (ruotare in verso opposto). Struttura identica a $(\mathbb Z,+)$, benché gli oggetti (numeri vs. rotazioni) siano completamente diversi.

### Definizione di gruppo

Un **gruppo** è una coppia $(G,*)$, con $G$ insieme non vuoto e $*:G\times G\to G$ un'operazione (una funzione, nel senso rigoroso del notebook 10.3), che soddisfa:

1. **Chiusura**: $a*b\in G$ per ogni $a,b\in G$ (già incorporata nella definizione di $*$ come funzione $G\times G\to G$);
2. **Associatività**: $(a*b)*c=a*(b*c)$ per ogni $a,b,c\in G$;
3. **Elemento neutro**: esiste $e\in G$ tale che $e*a=a*e=a$ per ogni $a\in G$;
4. **Inverso**: per ogni $a\in G$ esiste $a^{-1}\in G$ tale che $a*a^{-1}=a^{-1}*a=e$.

Se inoltre $a*b=b*a$ per ogni $a,b$ (**commutatività**), il gruppo si dice **abeliano**. Nota che la definizione **non richiede** la distributiva né un secondo elemento neutro: un gruppo ha una sola operazione, a differenza dei campi (che vedremo nel notebook 11.10, con due operazioni collegate dalla distributiva del notebook 0.5).

### Unicità dell'elemento neutro e dell'inverso

**Teorema.** In un gruppo $(G,*)$, l'elemento neutro è unico, e l'inverso di ogni elemento è unico.

**Dimostrazione (unicità del neutro).** Supponiamo che $e$ ed $e'$ siano entrambi elementi neutri. Allora $e=e*e'$ (perché $e'$ è neutro, applicando la definizione con $a=e$) $=e'$ (perché $e$ è neutro, applicando la definizione con $a=e'$). Dunque $e=e'$.

**Dimostrazione (unicità dell'inverso).** Supponiamo che $b$ e $c$ siano entrambi inversi di $a$: $a*b=e$ e $a*c=e$. Allora $b=b*e=b*(a*c)=(b*a)*c$ (associatività) $=e*c=c$. Dunque $b=c$. $\blacksquare$

Questa dimostrazione, puramente formale, non usa **nulla** oltre gli assiomi: è il prototipo di ogni dimostrazione in algebra astratta, dove non si può "controllare con un esempio numerico" ma solo manipolare gli assiomi.

### Esempi fondamentali

- $(\mathbb Z,+)$, $(\mathbb Q,+)$, $(\mathbb R,+)$: gruppi abeliani (neutro $0$, inverso $-a$).
- $(\mathbb Q\setminus\{0\},\times)$, $(\mathbb R\setminus\{0\},\times)$: gruppi abeliani (neutro $1$, inverso $1/a$) — nota che $(\mathbb Q,\times)$ **non** è un gruppo, perché $0$ non ha inverso moltiplicativo (l'esclusione di $0$ è essenziale).
- $(\mathbb R^n,+)$, lo spazio vettoriale del notebook 6.3: ogni spazio vettoriale è, "dimenticando" la moltiplicazione per scalari, un gruppo abeliano rispetto alla somma.
- Le matrici invertibili $n\times n$ con il prodotto righe per colonne (notebook 6.16): un gruppo **non abeliano** in generale (il prodotto tra matrici non commuta), indicato $GL_n(\mathbb R)$.
- Le simmetrie di una figura geometrica (rotazioni e riflessioni che la sovrappongono a se stessa), con la composizione: i **gruppi diedrali**.

### Proprietà elementari: la legge di cancellazione

**Teorema.** In un gruppo, $a*b=a*c\Rightarrow b=c$ (cancellazione a sinistra).

**Dimostrazione.** Componendo entrambi i membri a sinistra con $a^{-1}$: $a^{-1}*(a*b)=a^{-1}*(a*c)$. Per associatività, $(a^{-1}*a)*b=(a^{-1}*a)*c$, cioè $e*b=e*c$, cioè $b=c$. $\blacksquare$ Questa proprietà, banale per i numeri, è ciò che distingue un gruppo da una struttura più debole come un semplice **monoide** (che ha solo associatività e neutro, non l'inverso): senza l'inverso, la cancellazione può fallire, come accade per la moltiplicazione di matrici **non invertibili**.

````python
import sympy as sp
from sympy.combinatorics import Permutation, PermutationGroup

# Esempio: le rotazioni di un triangolo equilatero, come gruppo ciclico di ordine 3
# Rappresentiamo le rotazioni come permutazioni dei 3 vertici (0,1,2)
rot0 = Permutation([0, 1, 2])   # rotazione di 0 gradi (identità)
rot120 = Permutation([1, 2, 0]) # rotazione di 120 gradi
rot240 = Permutation([2, 0, 1]) # rotazione di 240 gradi

gruppo = PermutationGroup([rot120])
print("Ordine del gruppo generato dalla rotazione di 120°:", gruppo.order())
print("Elementi del gruppo:", [p.array_form for p in gruppo.elements])

# Verifica associativita' (banale per permutazioni, sempre vera) e chiusura
composizione = rot120 * rot120
print("\nComponendo due rotazioni di 120°, si ottiene:", composizione.array_form,
      "(equivalente a rotazione di 240°)", composizione.array_form == rot240.array_form)

# Verifica che ogni elemento ha un inverso nel gruppo
for p in gruppo.elements:
    inversa = p**-1
    print(f"Inversa di {p.array_form}: {inversa.array_form}, appartiene al gruppo? {inversa in gruppo}")
````

## Esercizio svolto e commentato

**Problema:** dimostra che $(\mathbb Z/n\mathbb Z,+)$ — l'insieme delle classi resto modulo $n$ (già incontrato nel notebook 10.3 come esempio di relazione di equivalenza) con la somma "modulo $n$" — è un gruppo abeliano.

**Ragionamento:**

**Passo 1 — definiamo l'operazione con precisione.** $\mathbb Z/n\mathbb Z=\{[0],[1],\ldots,[n-1]\}$, dove $[k]$ è la classe di equivalenza di $k$ rispetto alla congruenza modulo $n$ (notebook 10.3). Definiamo $[a]+[b]=[a+b]$. **Prima difficoltà, cruciale**: dobbiamo verificare che questa definizione sia **ben posta**, cioè non dipenda dal rappresentante scelto — se $[a]=[a']$ e $[b]=[b']$, deve risultare $[a+b]=[a'+b']$.

**Passo 2 — verifichiamo la buona posizione.** $[a]=[a']$ significa $a-a'=kn$ per qualche $k\in\mathbb Z$ (notebook 10.3); analogamente $b-b'=hn$. Sommando: $(a+b)-(a'+b')=(k+h)n$, quindi $a+b\equiv a'+b'\pmod n$, cioè $[a+b]=[a'+b']$: la somma è ben definita, indipendentemente dai rappresentanti.

**Passo 3 — verifichiamo gli assiomi di gruppo.** *Associatività*: $([a]+[b])+[c]=[a+b]+[c]=[(a+b)+c]=[a+(b+c)]=[a]+([b]+[c])$, ereditata dall'associatività di $+$ su $\mathbb Z$ (notebook 0.5). *Neutro*: $[0]$, perché $[a]+[0]=[a+0]=[a]$. *Inverso*: di $[a]$ è $[-a]$ (o equivalentemente $[n-a]$), perché $[a]+[-a]=[a-a]=[0]$. *Commutatività*: $[a]+[b]=[a+b]=[b+a]=[b]+[a]$, ereditata dalla commutatività di $+$ su $\mathbb Z$.

**Conclusione:** $(\mathbb Z/n\mathbb Z,+)$ è un gruppo abeliano, con $n$ elementi. **Il punto metodologico più importante**: ogni volta che si definisce un'operazione su classi di equivalenza (qui, resti modulo $n$), verificare la **buona posizione** è un passo dimostrativo indispensabile, non una formalità — è esattamente il tipo di controllo che, se omesso, produce "dimostrazioni" di fatti falsi su oggetti mal definiti.

````python
def somma_mod(a, b, n):
    return (a + b) % n

n = 5
elementi = list(range(n))

print(f"Tavola di composizione di (Z/{n}Z, +):")
print("   +", elementi)
for a in elementi:
    riga = [somma_mod(a, b, n) for b in elementi]
    print(f"   {a}", riga)

# Verifica degli assiomi di gruppo per Z/5Z
neutro = 0
print("\nElemento neutro:", neutro, "- verifica:", all(somma_mod(a, neutro, n) == a for a in elementi))

for a in elementi:
    inverso = (-a) % n
    print(f"Inverso di {a}: {inverso}, verifica: {somma_mod(a, inverso, n) == neutro}")
````

## Esercizi da risolvere

1. Verifica se $(\mathbb N,+)$ è un gruppo, motivando quale assioma fallisce (se ne fallisce uno).
2. Dimostra che in un gruppo $(a^{-1})^{-1}=a$ per ogni $a$ (usando solo gli assiomi e le proprietà di unicità dimostrate nella teoria).
3. Dimostra che in un gruppo $(a*b)^{-1}=b^{-1}*a^{-1}$ (attenzione all'ordine!) per ogni $a,b$.
4. *(più impegnativo)* Dimostra che se $(G,*)$ è un gruppo **finito** con $|G|=n$, e $a\in G$, allora esiste un intero positivo $k\le n$ tale che $a^k=e$ (dove $a^k=a*a*\cdots*a$, $k$ volte). Suggerimento: considera la successione $a,a^2,a^3,\ldots$ e usa il principio dei cassetti (se $G$ ha $n$ elementi, tra $a,a^2,\ldots,a^{n+1}$ due devono coincidere; poi usa la cancellazione).

**1.** $(\mathbb N,+)$ **non** è un gruppo: fallisce l'esistenza dell'inverso. Per esempio, $1\in\mathbb N$ non ha un opposto in $\mathbb N$ (dovrebbe essere $-1\notin\mathbb N$). Chiusura, associatività e neutro ($0$, se si include in $\mathbb N$) valgono, ma senza inverso la struttura è solo un **monoide**.

**2.** Per definizione, $a^{-1}$ è l'elemento che soddisfa $a*a^{-1}=a^{-1}*a=e$. Questa stessa uguaglianza, letta "al contrario" (scambiando i ruoli), dice che $a$ è **un** inverso di $a^{-1}$: $a^{-1}*a=a*a^{-1}=e$. Per l'unicità dell'inverso (dimostrata nella teoria), l'inverso di $a^{-1}$ è unico, quindi $(a^{-1})^{-1}=a$.

**3.** Verifichiamo che $b^{-1}*a^{-1}$ soddisfa la definizione di inverso di $a*b$: $(a*b)*(b^{-1}*a^{-1})=a*(b*b^{-1})*a^{-1}$ (associatività) $=a*e*a^{-1}=a*a^{-1}=e$. Analogamente, $(b^{-1}*a^{-1})*(a*b)=b^{-1}*(a^{-1}*a)*b=b^{-1}*e*b=b^{-1}*b=e$. Per l'unicità dell'inverso, $(a*b)^{-1}=b^{-1}*a^{-1}$.

**4.** Consideriamo gli $n+1$ elementi $a,a^2,\ldots,a^{n+1}$ (tutti in $G$, che ha solo $n$ elementi distinti): per il principio dei cassetti, almeno due coincidono, diciamo $a^i=a^j$ con $1\le i<j\le n+1$. Moltiplicando entrambi i membri per $(a^{-1})^i=(a^i)^{-1}$ (ben definito per l'esercizio 3, generalizzato): $e=a^{j-i}$, con $k=j-i$ che soddisfa $1\le k\le n$ (perché $1\le i<j\le n+1$). Dunque $a^k=e$ con $k\le n$.

Verifica computazionale (esercizio 4, su un gruppo concreto):
```python
from sympy.combinatorics import Permutation, PermutationGroup

rot120 = Permutation([1, 2, 0])
gruppo = PermutationGroup([rot120])
n = gruppo.order()
print("Ordine del gruppo:", n)

potenza = rot120
k = 1
while potenza != Permutation([0, 1, 2]):  # finche' non torniamo all'identita'
    potenza = potenza * rot120
    k += 1
print(f"La prima potenza che restituisce l'identita' e' k={k} (atteso <= {n})")
```

## Riepilogo finale

- Un **gruppo** è un insieme con un'operazione associativa, dotata di elemento neutro e inverso per ogni elemento — la struttura comune a numeri, vettori, matrici invertibili, simmetrie geometriche.
- Neutro e inverso sono sempre **unici**, e la **legge di cancellazione** vale sempre in un gruppo: entrambi dimostrati dai soli assiomi, senza riferimento a esempi specifici.
- $(\mathbb Z/n\mathbb Z,+)$ è l'esempio archetipico di gruppo finito, costruito su classi di equivalenza (notebook 10.3) — richiede di verificare che l'operazione sia **ben posta**.
- L'algebra astratta guadagna potenza dimostrando teoremi **una volta per tutte** sui gruppi, applicabili automaticamente a ogni struttura che ne soddisfi gli assiomi.
