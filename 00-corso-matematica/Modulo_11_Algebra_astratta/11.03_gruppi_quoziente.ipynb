# 11.3 — Gruppi quoziente

**Prerequisiti:** 11.2 Sottogruppi, 10.3 Insiemi, relazioni, funzioni (rigorosi).


## Perché ci interessa

Nell'esercizio svolto del notebook 11.1 abbiamo costruito $\mathbb Z/n\mathbb Z$ come gruppo delle classi resto, verificando "a mano" che la somma di classi fosse ben definita. Il notebook 11.2 ha mostrato che le classi laterali $aH$ di un sottogruppo $H\le G$ partizionano sempre $G$ — ma **non sempre** è possibile trasformare l'insieme delle classi laterali in un gruppo, componendole tra loro. Questo notebook individua esattamente la condizione che lo rende possibile — la **normalità** del sottogruppo — e costruisce il **gruppo quoziente** $G/H$, generalizzando $\mathbb Z/n\mathbb Z$ a un contesto arbitrario, anche non abeliano.

## Teoria, passo dopo passo

### Il problema: comporre classi laterali

Vorremmo definire, sull'insieme $G/H=\{aH:a\in G\}$ delle classi laterali, un'operazione $(aH)*(bH)=(a*b)H$. Ma, come nell'esercizio del notebook 11.1, questa definizione usa i **rappresentanti** $a,b$: dobbiamo verificare che sia **ben posta**, cioè che non dipenda dalla scelta dei rappresentanti. Se $aH=a'H$ e $bH=b'H$, vorremmo $(ab)H=(a'b')H$.

### Sottogruppi normali

Un sottogruppo $H\le G$ è **normale** (si scrive $H\trianglelefteq G$) se $aHa^{-1}=H$ per ogni $a\in G$ — equivalentemente, $aH=Ha$ per ogni $a$ (la classe laterale sinistra coincide con quella destra). In un gruppo **abeliano**, ogni sottogruppo è automaticamente normale, perché $ah=ha$ sempre.

**Teorema.** Se $H\trianglelefteq G$, l'operazione $(aH)*(bH)=(ab)H$ su $G/H$ è ben definita.

**Dimostrazione.** Supponiamo $aH=a'H$ e $bH=b'H$: per la caratterizzazione delle classi laterali (notebook 11.2), $a^{-1}a'\in H$ e $b^{-1}b'\in H$. Vogliamo $(ab)H=(a'b')H$, cioè $(ab)^{-1}(a'b')\in H$. Calcoliamo: $(ab)^{-1}(a'b')=b^{-1}a^{-1}a'b'$. Scriviamo $a^{-1}a'=h_1\in H$: l'espressione diventa $b^{-1}h_1b'=b^{-1}h_1b\cdot b^{-1}b'$. Per **normalità** di $H$, $b^{-1}h_1b\in H$ (perché $H=b^{-1}Hb$, conseguenza di $bH=Hb$ per ogni $b$). E $b^{-1}b'\in H$ per ipotesi. Il prodotto di due elementi di $H$ è in $H$ (chiusura): $b^{-1}h_1b\cdot b^{-1}b'\in H$. Dunque $(ab)^{-1}(a'b')\in H$, cioè $(ab)H=(a'b')H$. $\blacksquare$

**Perché la normalità è essenziale**: il passaggio cruciale, $b^{-1}h_1b\in H$, richiede di poter "spostare" un elemento di $H$ attraverso la coniugazione per $b$ **restando dentro $H$** — esattamente la definizione di normalità. Senza questa proprietà, la dimostrazione si blocca, e infatti si può costruire un controesempio esplicito con un sottogruppo non normale dove l'operazione risulta mal definita.

### Il gruppo quoziente $G/H$

**Teorema.** Se $H\trianglelefteq G$, allora $(G/H,*)$ è un gruppo, detto **gruppo quoziente**.

**Dimostrazione (verifica degli assiomi).** *Associatività*: $((aH)*(bH))*(cH)=(ab)H*(cH)=((ab)c)H=(a(bc))H$ (associatività in $G$) $=(aH)*((bc)H)=(aH)*((bH)*(cH))$. *Neutro*: $eH=H$, perché $(aH)*(eH)=(ae)H=aH$. *Inverso*: di $aH$ è $a^{-1}H$, perché $(aH)*(a^{-1}H)=(aa^{-1})H=eH=H$. $\blacksquare$

L'esempio $\mathbb Z/n\mathbb Z$ del notebook 11.1 è ora visto come caso particolare: $H=n\mathbb Z\trianglelefteq(\mathbb Z,+)$ (normale automaticamente, essendo $\mathbb Z$ abeliano), e $G/H=\mathbb Z/n\mathbb Z$ è precisamente il gruppo quoziente qui costruito in generale.

### Un esempio di sottogruppo NON normale

Nel gruppo diedrale $D_3$ (simmetrie di un triangolo, notebook 11.2, ordine 6), il sottogruppo generato da una singola riflessione $H=\{e,r\}$ (dove $r$ è una riflessione, $r^2=e$) ha ordine 2, ma **non** è normale: coniugando $r$ con una rotazione $\rho$ si ottiene $\rho r\rho^{-1}$, che è una riflessione **diversa** da $r$ (geometricamente, ruotare l'asse di riflessione), quindi $\rho H\rho^{-1}\ne H$. In questo caso, $G/H$ non eredita una struttura di gruppo ben definita — il tentativo di comporre le classi laterali dipenderebbe dalla scelta dei rappresentanti.

````python
from sympy.combinatorics import Permutation, PermutationGroup

# D3: rotazioni + riflessioni del triangolo
rot = Permutation([1, 2, 0])
rifl = Permutation([0, 2, 1])
D3 = PermutationGroup([rot, rifl])
print("Ordine di D3:", D3.order())

# Sottogruppo delle rotazioni (normale, indice 2)
H_rotazioni = PermutationGroup([rot])
print("H (rotazioni) e' normale in D3?", D3.is_normal(H_rotazioni))

# Sottogruppo generato da una singola riflessione (NON normale)
H_riflessione = PermutationGroup([rifl])
print("H (una riflessione) e' normale in D3?", D3.is_normal(H_riflessione))

# Verifica esplicita: coniughiamo la riflessione con la rotazione
coniugata = rot * rifl * rot**-1
print("\nRiflessione originale:", rifl.array_form)
print("Riflessione coniugata (rot * rifl * rot^-1):", coniugata.array_form)
print("Sono la stessa permutazione?", coniugata == rifl)
````

## Esercizio svolto e commentato

**Problema:** dimostra che **ogni sottogruppo di indice 2** è normale (un criterio pratico molto usato, che spiega perché $H_{\text{rotazioni}}$ nell'esempio computazionale sopra, di indice $6/3=2$, è risultato normale).

**Ragionamento:**

**Passo 1 — descriviamo le classi laterali quando l'indice è 2.** Se $[G:H]=2$, ci sono esattamente due classi laterali sinistre: $H$ stesso (la classe di $e$) e un'unica altra classe $aH$, per un qualsiasi $a\notin H$. Poiché le classi laterali partizionano $G$ (notebook 11.2), necessariamente $aH=G\setminus H$ (il complementare di $H$).

**Passo 2 — lo stesso vale per le classi laterali destre.** Con lo stesso argomento applicato a $Ha$: ci sono due classi laterali destre, $H$ e $Ha=G\setminus H$ (lo stesso complementare, perché ogni elemento non in $H$ deve stare nell'unica altra classe destra).

**Passo 3 — confrontiamo.** Se $a\in H$: $aH=H=Ha$ banalmente. Se $a\notin H$: $aH=G\setminus H$ (passo 1) e $Ha=G\setminus H$ (passo 2): quindi $aH=Ha$ anche in questo caso.

**Conclusione:** per ogni $a\in G$, $aH=Ha$: $H$ è normale. **Il motivo intuitivo**: con solo due classi laterali, non c'è "spazio" per una classe sinistra e una destra diverse — il complementare di $H$ è un insieme unico, forzato a coincidere sia da sinistra sia da destra. Questo criterio è particolarmente utile perché evita di dover verificare esplicitamente la coniugazione per ogni elemento del gruppo.

````python
from sympy.combinatorics import Permutation, PermutationGroup

rot = Permutation([1, 2, 0])
rifl = Permutation([0, 2, 1])
D3 = PermutationGroup([rot, rifl])

H = PermutationGroup([rot])  # indice 2 in D3 (ordine 3 su 6)
indice = D3.order() // H.order()
print("Indice di H in D3:", indice)
print("H e' normale (verificato dal criterio 'indice 2')?", D3.is_normal(H))

# Verifica diretta: classi laterali sinistre e destre coincidono per ogni elemento
for g in D3.elements:
    sinistra = sorted([(g*h).array_form for h in H.elements])
    destra = sorted([(h*g).array_form for h in H.elements])
    print(f"g={g.array_form}: gH == Hg? {sinistra == destra}")
````

## Riepilogo finale

- Un sottogruppo $H\trianglelefteq G$ è **normale** se $aHa^{-1}=H$ per ogni $a$: condizione necessaria e sufficiente perché l'operazione $(aH)(bH)=(ab)H$ su $G/H$ sia **ben definita**.
- In un gruppo **abeliano**, ogni sottogruppo è automaticamente normale; un **sottogruppo di indice 2** è sempre normale, qualunque sia il gruppo.
- Il **gruppo quoziente** $G/H$ generalizza $\mathbb Z/n\mathbb Z$: eredita associatività, neutro ($H$ stesso) e inverso ($a^{-1}H$) direttamente da $G$.
- Non tutti i sottogruppi sono normali: nel gruppo diedrale $D_3$, il sottogruppo generato da una singola riflessione non lo è.
