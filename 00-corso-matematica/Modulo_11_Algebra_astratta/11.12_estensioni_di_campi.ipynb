## Perché ci interessa

Abbiamo costruito $\mathbb C=\mathbb R[x]/(x^2+1)$ (notebook 11.8) e verificato che $\mathbb Q(\sqrt2)$ è un campo (notebook 11.10): entrambi sono esempi di **estensione di campo** — un campo più grande costruito "aggiungendo" un elemento a un campo di partenza. Questo notebook sistematizza la costruzione, mostrando che ogni estensione semplice è, in un senso preciso, uno **spazio vettoriale** sul campo di partenza (collegandosi al notebook 6.9), con una dimensione ben definita legata al grado del polinomio minimo dell'elemento aggiunto. Questo è l'ultimo strumento prima della teoria di Galois (notebook 11.13), che userà proprio questa nozione di grado per spiegare perché alcune equazioni non si possono risolvere per radicali.

## Teoria, passo dopo passo

### Elementi algebrici e trascendenti

Sia $F\subseteq E$ un'estensione di campi (cioè $F$ è un sottocampo di $E$). Un elemento $\alpha\in E$ è **algebrico su $F$** se esiste un polinomio non nullo $p(x)\in F[x]$ con $p(\alpha)=0$; altrimenti è **trascendente**. Per esempio, $\sqrt2\in\mathbb R$ è algebrico su $\mathbb Q$ (radice di $x^2-2$); $i\in\mathbb C$ è algebrico su $\mathbb R$ (radice di $x^2+1$); $\pi$ è **trascendente** su $\mathbb Q$ (un fatto profondo, dimostrato da Lindemann nel 1882, che esula da questo corso ma la cui esistenza è garantita — con un argomento di cardinalità — dal notebook 10.5: i numeri algebrici sono numerabili, essendo radici di polinomi a coefficienti in $\mathbb Q$ che sono numerabili, mentre $\mathbb R$ non lo è, quindi "quasi tutti" i reali sono trascendenti).

### Il polinomio minimo

Se $\alpha$ è algebrico su $F$, tra tutti i polinomi non nulli di $F[x]$ che si annullano in $\alpha$, ne esiste uno di **grado minimo** (per il principio del buon ordinamento, notebook 10.4), detto **polinomio minimo** $m_\alpha(x)$, normalizzato a coefficiente direttore $1$.

**Teorema.** Il polinomio minimo $m_\alpha(x)$ è **irriducibile** su $F$, ed è unico con questa proprietà di minimalità.

**Dimostrazione (irriducibilità, per assurdo).** Supponiamo $m_\alpha=gh$ con $\deg g,\deg h<\deg m_\alpha$. Allora $0=m_\alpha(\alpha)=g(\alpha)h(\alpha)$; essendo $E$ un campo (dominio integrale, notebook 11.9), $g(\alpha)=0$ oppure $h(\alpha)=0$ — ma entrambi hanno grado minore di $m_\alpha$, contraddicendo la minimalità di $m_\alpha$. $\blacksquare$

**Esempi**: il polinomio minimo di $\sqrt2$ su $\mathbb Q$ è $x^2-2$ (grado 2: nessun polinomio di grado 1, cioè $x-\sqrt2$, ha coefficienti razionali). Il polinomio minimo di $i$ su $\mathbb R$ è $x^2+1$.

### Estensioni semplici come spazi vettoriali

Dato $\alpha$ algebrico su $F$ con polinomio minimo di grado $n$, definiamo $F(\alpha)$ come il più piccolo campo contenente $F$ e $\alpha$.

**Teorema.** $F(\alpha)\cong F[x]/(m_\alpha(x))$, e $F(\alpha)$ è uno spazio vettoriale su $F$ di dimensione $n=\deg m_\alpha$, con base $\{1,\alpha,\alpha^2,\ldots,\alpha^{n-1}\}$.

**Dimostrazione (isomorfismo, tramite valutazione).** L'omomorfismo di valutazione $\operatorname{ev}_\alpha:F[x]\to E$, $p(x)\mapsto p(\alpha)$, ha nucleo $(m_\alpha(x))$: se $p(\alpha)=0$, dividendo $p$ per $m_\alpha$ (divisione euclidea, notebook 11.11) si ottiene $p=qm_\alpha+r$ con $\deg r<\deg m_\alpha$; valutando in $\alpha$: $0=p(\alpha)=q(\alpha)m_\alpha(\alpha)+r(\alpha)=r(\alpha)$, ma $m_\alpha$ era di grado **minimo** tra i polinomi che si annullano in $\alpha$, quindi $r$ (di grado inferiore) deve essere il polinomio nullo: $p$ è un multiplo di $m_\alpha$. Per il primo teorema di isomorfismo (notebook 11.8), $F[x]/(m_\alpha)\cong\operatorname{Im}(\operatorname{ev}_\alpha)=F(\alpha)$ (l'immagine è proprio $F(\alpha)$, il più piccolo campo contenente $F$ e $\alpha$: qui si usa che $m_\alpha$ irriducibile rende $F[x]/(m_\alpha)$ un **campo**, non solo un anello — perché in un PID come $F[x]$, il quoziente per un ideale generato da un elemento irriducibile è un campo, in perfetta analogia con $\mathbb Z/p\mathbb Z$ per $p$ primo).

**Dimensione**: ogni classe di $F[x]/(m_\alpha)$ ha, per la divisione euclidea, un unico rappresentante di grado $<n$, cioè della forma $c_0+c_1x+\cdots+c_{n-1}x^{n-1}$: questi rappresentanti corrispondono, sotto l'isomorfismo, a $c_0+c_1\alpha+\cdots+c_{n-1}\alpha^{n-1}$, mostrando che $\{1,\alpha,\ldots,\alpha^{n-1}\}$ genera $F(\alpha)$ come spazio vettoriale su $F$ (nel senso del notebook 6.7); l'indipendenza lineare (notebook 6.6) segue dalla minimalità di $m_\alpha$ (una combinazione lineare nulla non banale darebbe un polinomio di grado $<n$ che si annulla in $\alpha$). $\blacksquare$

Il numero $n=\dim_FF(\alpha)$ si indica $[F(\alpha):F]$, il **grado** dell'estensione. Per $\mathbb C=\mathbb R(i)$: $[\mathbb C:\mathbb R]=2$, con base $\{1,i\}$ — esattamente la base usata fin dal notebook 9.1 per la forma cartesiana $a+bi$, ora vista come base di uno spazio vettoriale di dimensione 2. Per $\mathbb Q(\sqrt2)$: $[\mathbb Q(\sqrt2):\mathbb Q]=2$, con base $\{1,\sqrt2\}$.

````python
import sympy as sp

x = sp.symbols('x')

# Polinomio minimo di sqrt(2) su Q
alpha = sp.sqrt(2)
min_poly = sp.minimal_polynomial(alpha, x)
print(f"Polinomio minimo di sqrt(2) su Q: {min_poly}, grado {sp.degree(min_poly, x)}")

# Polinomio minimo di una radice piu' complessa: 1+sqrt(2)
beta = 1 + sp.sqrt(2)
min_poly_beta = sp.minimal_polynomial(beta, x)
print(f"Polinomio minimo di 1+sqrt(2) su Q: {min_poly_beta}, grado {sp.degree(min_poly_beta, x)}")

# Polinomio minimo della radice cubica di 2 (estensione di grado 3)
gamma = sp.Rational(2)**sp.Rational(1,3)
min_poly_gamma = sp.minimal_polynomial(gamma, x)
print(f"Polinomio minimo di 2^(1/3) su Q: {min_poly_gamma}, grado {sp.degree(min_poly_gamma, x)}")
print("-> [Q(2^(1/3)):Q] = 3, base {1, 2^(1/3), 2^(2/3)}")
````

## Esercizio svolto e commentato

**Problema:** determina il grado $[\mathbb Q(\sqrt2,\sqrt3):\mathbb Q]$ (l'estensione ottenuta aggiungendo **entrambi** $\sqrt2$ e $\sqrt3$ a $\mathbb Q$), e trova una base.

**Ragionamento:**

**Passo 1 — costruiamo l'estensione in due passi.** $\mathbb Q(\sqrt2,\sqrt3)=\mathbb Q(\sqrt2)(\sqrt3)$: prima estendiamo $\mathbb Q$ con $\sqrt2$ (grado 2, base $\{1,\sqrt2\}$, notebook 11.10), poi estendiamo $\mathbb Q(\sqrt2)$ con $\sqrt3$.

**Passo 2 — determiniamo il polinomio minimo di $\sqrt3$ **su $\mathbb Q(\sqrt2)$** (non su $\mathbb Q$!).** $\sqrt3$ è radice di $x^2-3$, che ha coefficienti in $\mathbb Q\subset\mathbb Q(\sqrt2)$. È ancora irriducibile su $\mathbb Q(\sqrt2)$? Dobbiamo verificare che $\sqrt3\notin\mathbb Q(\sqrt2)$: se fosse $\sqrt3=a+b\sqrt2$ con $a,b\in\mathbb Q$, elevando al quadrato $3=a^2+2b^2+2ab\sqrt2$; se $ab\ne0$, questo darebbe $\sqrt2=\dfrac{3-a^2-2b^2}{2ab}\in\mathbb Q$, assurdo ($\sqrt2$ irrazionale, notebook 10.6); se $ab=0$, o $a=0$ (allora $3=2b^2$, cioè $b^2=3/2$, ma nessun razionale ha quadrato $3/2$) o $b=0$ (allora $3=a^2$, ma nessun razionale ha quadrato $3$): in ogni caso, contraddizione. Dunque $\sqrt3\notin\mathbb Q(\sqrt2)$, quindi $x^2-3$ resta irriducibile su $\mathbb Q(\sqrt2)$ (nessuna radice in quel campo, e per un polinomio di secondo grado questo basta per l'irriducibilità, come nell'esercizio 4 del notebook 11.11).

**Passo 3 — applichiamo il teorema del grado moltiplicativamente.** $[\mathbb Q(\sqrt2)(\sqrt3):\mathbb Q(\sqrt2)]=2$ (grado del polinomio minimo $x^2-3$ su $\mathbb Q(\sqrt2)$). Per la **legge delle torri** (un fatto generale, non dimostrato qui in dettaglio, ma intuitivo: le basi si "moltiplicano" — se $\{1,\sqrt2\}$ è base di $\mathbb Q(\sqrt2)$ su $\mathbb Q$ e $\{1,\sqrt3\}$ è base di $\mathbb Q(\sqrt2,\sqrt3)$ su $\mathbb Q(\sqrt2)$, allora i 4 prodotti $\{1,\sqrt2,\sqrt3,\sqrt6\}$ formano una base sul campo più piccolo $\mathbb Q$): $[\mathbb Q(\sqrt2,\sqrt3):\mathbb Q]=[\mathbb Q(\sqrt2,\sqrt3):\mathbb Q(\sqrt2)]\times[\mathbb Q(\sqrt2):\mathbb Q]=2\times2=4$.

**Conclusione:** $[\mathbb Q(\sqrt2,\sqrt3):\mathbb Q]=4$, con base $\{1,\sqrt2,\sqrt3,\sqrt6\}$. **Il valore di questo esempio**: mostra che le estensioni si compongono **moltiplicativamente** (non additivamente), un fatto che nella teoria di Galois (notebook 11.13) diventa lo strumento per contare esattamente quante "torri" di estensioni servono per esprimere le radici di un'equazione — e, quando nessuna torre finita basta, per dimostrare che l'equazione non è risolvibile per radicali.

````python
import sympy as sp

x = sp.symbols('x')

# Verifica che {1, sqrt2, sqrt3, sqrt6} sono linearmente indipendenti su Q
# (equivalente a verificare che il grado dell'estensione e' davvero 4)
alpha = sp.sqrt(2) + sp.sqrt(3)
min_poly = sp.minimal_polynomial(alpha, x)
print(f"Polinomio minimo di sqrt(2)+sqrt(3) su Q: {min_poly}")
print(f"Grado: {sp.degree(min_poly, x)} (conferma [Q(sqrt2,sqrt3):Q] = 4)")

# Verifica numerica che sqrt(3) non e' in Q(sqrt2) (nessuna combinazione razionale a+b*sqrt2 da' sqrt3)
import itertools
sqrt2, sqrt3 = float(sp.sqrt(2)), float(sp.sqrt(3))
trovato = False
for num_a in range(-20, 21):
    for den_a in range(1, 10):
        for num_b in range(-20, 21):
            for den_b in range(1, 10):
                a, b = num_a/den_a, num_b/den_b
                if abs(a + b*sqrt2 - sqrt3) < 1e-9:
                    trovato = True
print("Trovata rappresentazione razionale a+b*sqrt2 = sqrt3 (ricerca approssimata)?", trovato)
````

## Riepilogo finale

- Un elemento $\alpha$ è **algebrico su $F$** se è radice di qualche polinomio non nullo in $F[x]$; il suo **polinomio minimo** è irriducibile e unico.
- $F(\alpha)\cong F[x]/(m_\alpha(x))$ è uno spazio vettoriale su $F$ di **dimensione** $n=\deg m_\alpha$, con base $\{1,\alpha,\ldots,\alpha^{n-1}\}$: il **grado** dell'estensione $[F(\alpha):F]$.
- $[\mathbb C:\mathbb R]=2$, $[\mathbb Q(\sqrt2):\mathbb Q]=2$: esempi concreti della costruzione generale.
- La **legge delle torri** ($[E:F]=[E:K][K:F]$ per estensioni concatenate $F\subseteq K\subseteq E$) permette di calcolare gradi di estensioni composte, come $[\mathbb Q(\sqrt2,\sqrt3):\mathbb Q]=4$.
