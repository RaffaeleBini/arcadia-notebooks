## Perché ci interessa

Nel notebook 6.4 abbiamo studiato i **sottospazi** di uno spazio vettoriale: sottoinsiemi che ereditano la struttura dello spazio ambiente. L'idea si generalizza naturalmente ai gruppi: un **sottogruppo** è un sottoinsieme che, con la stessa operazione, è esso stesso un gruppo. Esempi concreti sono ovunque: i numeri pari $2\mathbb Z$ dentro $\mathbb Z$, le rotazioni (senza riflessioni) dentro il gruppo di simmetria di un poligono, le matrici con determinante 1 dentro $GL_n(\mathbb R)$. Questo notebook, oltre a caratterizzare i sottogruppi con un criterio pratico, dimostra uno dei risultati più eleganti e sorprendenti della teoria elementare dei gruppi: il **teorema di Lagrange**, che lega la dimensione (l'ordine) di un sottogruppo a quella del gruppo — con conseguenze aritmetiche immediate.

## Teoria, passo dopo passo

### Definizione e criterio del sottogruppo

Un sottoinsieme $H\subseteq G$ è un **sottogruppo** di $(G,*)$ se $(H,*)$ è esso stesso un gruppo — cioè se $H$ contiene il neutro $e$, è chiuso rispetto a $*$, e ogni suo elemento ha inverso **in $H$**. Verificare tutti gli assiomi da zero è ridondante (l'associatività è ereditata automaticamente da $G$): basta un **criterio del sottogruppo**.

**Teorema (criterio del sottogruppo).** $H\subseteq G$ non vuoto è un sottogruppo $\iff$ per ogni $a,b\in H$, $a*b^{-1}\in H$.

**Dimostrazione ($\Rightarrow$).** Se $H$ è un sottogruppo, per definizione contiene gli inversi e è chiuso, quindi $a,b\in H\Rightarrow b^{-1}\in H\Rightarrow a*b^{-1}\in H$.

**Dimostrazione ($\Leftarrow$).** Supponiamo la condizione. Poiché $H\ne\emptyset$, esiste $a\in H$; applicando la condizione con $b=a$: $a*a^{-1}=e\in H$ (il neutro è in $H$). Applicando la condizione con $a=e,b$ generico in $H$: $e*b^{-1}=b^{-1}\in H$ (ogni elemento ha inverso in $H$). Infine, dati $a,b\in H$: sappiamo ora $b^{-1}\in H$, quindi applicando la condizione a $a$ e $b^{-1}$: $a*(b^{-1})^{-1}=a*b\in H$ (chiusura). Tutti gli assiomi verificati (l'associatività è automatica, essendo $H\subseteq G$): $H$ è un sottogruppo. $\blacksquare$

### Esempi

- $n\mathbb Z=\{nk:k\in\mathbb Z\}$ (i multipli di $n$) è un sottogruppo di $(\mathbb Z,+)$: se $a=nk_1,b=nk_2$, allora $a-b=n(k_1-k_2)\in n\mathbb Z$ (qui l'operazione è $+$, quindi il criterio diventa $a-b\in H$).
- $\{0°,120°,240°\}$ (le rotazioni) è un sottogruppo del gruppo diedrale delle simmetrie di un triangolo (che include anche le riflessioni).
- $SL_n(\mathbb R)$ (matrici con determinante $1$) è un sottogruppo di $GL_n(\mathbb R)$: se $\det A=\det B=1$, allora $\det(AB^{-1})=\det A\cdot(\det B)^{-1}=1$ (proprietà del determinante, notebook 6.12).
- $\{e\}$ (il **sottogruppo banale**) e $G$ stesso sono sempre sottogruppi di $G$.

### Classi laterali

Fissato un sottogruppo $H\le G$ e un elemento $a\in G$, la **classe laterale sinistra** è $aH=\{a*h:h\in H\}$. Definiamo una relazione su $G$: $x\sim y\iff x^{-1}*y\in H$. **Verifichiamo che è una relazione di equivalenza** (nel senso del notebook 10.3): *riflessiva* ($x^{-1}*x=e\in H$); *simmetrica* (se $x^{-1}y\in H$, il suo inverso $(x^{-1}y)^{-1}=y^{-1}x\in H$ per la chiusura rispetto agli inversi); *transitiva* (se $x^{-1}y\in H$ e $y^{-1}z\in H$, il loro prodotto $(x^{-1}y)(y^{-1}z)=x^{-1}z\in H$ per la chiusura). Le classi di equivalenza di questa relazione sono **esattamente** le classi laterali $xH$: infatti $x\sim y\iff x^{-1}y\in H\iff y\in xH$.

**Osservazione chiave**: ogni classe laterale $aH$ ha **la stessa cardinalità** di $H$: la funzione $h\mapsto a*h$ è una biiezione $H\to aH$ (iniettiva per cancellazione, notebook 11.1; suriettiva per definizione di $aH$).

### Il teorema di Lagrange

**Teorema (Lagrange).** Se $G$ è un gruppo **finito** e $H\le G$ un sottogruppo, allora $|H|$ divide $|G|$.

**Dimostrazione.** Per la teoria appena vista, le classi laterali $\{aH:a\in G\}$ sono le classi di equivalenza di una relazione su $G$, quindi **partizionano** $G$ (notebook 10.3: classi di equivalenza sempre disgiunte o coincidenti, e la loro unione è tutto $G$). Ogni classe laterale ha cardinalità $|H|$ (per l'osservazione sopra). Se ci sono $k$ classi laterali distinte, allora $|G|=k\cdot|H|$ (l'insieme $G$ si spezza in $k$ blocchi disgiunti, ciascuno di dimensione $|H|$ — un conteggio analogo a quello delle partizioni finite del notebook 10.3). Dunque $|H|$ divide $|G|$. $\blacksquare$

Il numero $k=|G|/|H|$ si chiama **indice** di $H$ in $G$. Questo teorema, dimostrato con un argomento puramente insiemistico (partizione in classi laterali di uguale cardinalità), ha conseguenze aritmetiche potenti: per esempio, un gruppo di ordine primo $p$ non può avere sottogruppi non banali (perché gli unici divisori di $p$ sono $1$ e $p$), un fatto che userà il notebook 11.5 per classificare i gruppi di ordine primo.

````python
from sympy.combinatorics import Permutation, PermutationGroup
from sympy import divisors

# Il gruppo diedrale D3 (simmetrie di un triangolo): 3 rotazioni + 3 riflessioni, ordine 6
rot = Permutation([1, 2, 0])       # rotazione di 120
rifl = Permutation([0, 2, 1])      # una riflessione

D3 = PermutationGroup([rot, rifl])
print("Ordine di D3:", D3.order())

# Il sottogruppo delle sole rotazioni (ciclico, ordine 3)
H = PermutationGroup([rot])
print("Ordine del sottogruppo H (rotazioni):", H.order())
print("|H| divide |D3|?", D3.order() % H.order() == 0)
print("Indice [D3:H] =", D3.order() // H.order())

print("\nDivisori di |D3| (possibili ordini di sottogruppi, per Lagrange):", divisors(D3.order()))
````

## Esercizio svolto e commentato

**Problema:** dimostra che ogni gruppo $G$ di ordine primo $p$ è **ciclico** (cioè generato da un singolo elemento: $G=\{e,a,a^2,\ldots,a^{p-1}\}$ per qualche $a\in G$).

**Ragionamento:**

**Passo 1 — scegliamo un elemento non banale.** Poiché $p\ge2$ (essendo primo), $G$ contiene un elemento $a\ne e$.

**Passo 2 — consideriamo il sottogruppo generato da $a$.** Sia $\langle a\rangle=\{a^k:k\in\mathbb Z\}$ (le potenze di $a$, incluse quelle negative $a^{-k}=(a^{-1})^k$): si verifica facilmente, con il criterio del sottogruppo, che $\langle a\rangle$ è un sottogruppo di $G$.

**Passo 3 — applichiamo Lagrange.** $|\langle a\rangle|$ divide $|G|=p$. Poiché $p$ è primo, i suoi unici divisori sono $1$ e $p$.

**Passo 4 — escludiamo $|\langle a\rangle|=1$.** Se $|\langle a\rangle|=1$, allora $\langle a\rangle=\{e\}$, ma $a\in\langle a\rangle$ e $a\ne e$ per costruzione (passo 1) — contraddizione.

**Conclusione:** deve essere $|\langle a\rangle|=p=|G|$, quindi $\langle a\rangle=G$ (un sottoinsieme di $G$ con la stessa cardinalità finita di $G$ deve coincidere con $G$): $G$ è generato dal singolo elemento $a$, cioè è ciclico. **Il valore di questo risultato**: classifica **completamente** i gruppi di ordine primo (per esempio ordine $2,3,5,7,11,\ldots$): sono tutti "uguali" (isomorfi, concetto che formalizzeremo nel notebook 11.4) a $(\mathbb Z/p\mathbb Z,+)$ — non esistono altre strutture di gruppo possibili con quell'ordine, un fatto tutt'altro che ovvio dimostrato in poche righe grazie a Lagrange.

````python
from sympy.combinatorics import Permutation, PermutationGroup, CyclicGroup

# Verifica per p=5: ogni gruppo di ordine 5 e' ciclico
G = CyclicGroup(5)
print("Ordine di G:", G.order())
print("G e' ciclico?", G.is_cyclic)

# Generiamo il sottogruppo ciclico a partire da un elemento non identita'
elementi = list(G.elements)
identita = G.identity
a = [e for e in elementi if e != identita][0]

sottogruppo_generato = PermutationGroup([a])
print(f"\nOrdine del sottogruppo generato da un elemento non identico: {sottogruppo_generato.order()}")
print("Coincide con l'intero gruppo G?", sottogruppo_generato.order() == G.order())
````

**4.** Sia $a$ un elemento di ordine $2p$: allora $\langle a\rangle=\{e,a,a^2,\ldots,a^{2p-1}\}$ ha $2p$ elementi, quindi $\langle a\rangle=G$ (essendo $|G|=2p$): $G$ è ciclico, generato da $a$. Consideriamo $b=a^2$: l'ordine di $b$ è il più piccolo $k$ con $(a^2)^k=e$, cioè $a^{2k}=e$; poiché l'ordine di $a$ è $2p$, questo richiede $2p\mid2k$, cioè $p\mid k$, e il più piccolo tale $k$ positivo è $k=p$ (dato che $p$ è primo, non ci sono divisori intermedi). Dunque $b=a^2$ ha ordine $p$, e $\langle a^2\rangle=\{e,a^2,a^4,\ldots,a^{2(p-1)}\}$ è un sottogruppo di ordine $p$. Per l'unicità: qualunque altro sottogruppo di ordine $p$ dovrebbe, per Lagrange, avere ordine che divide $2p$; essendo $G$ ciclico generato da $a$, ogni suo sottogruppo è della forma $\langle a^d\rangle$ per qualche divisore $d$ di $2p$, e l'unico che dà ordine $p$ è $d=2$: il sottogruppo di ordine $p$ è unico.

def sottogruppo_generato(g, n):
    elementi = set()
    x = 0
    while x not in elementi:
        elementi.add(x)
        x = (x + g) % n
    return sorted(elementi)

n = 12

    g = n // d
    sg = sottogruppo_generato(g, n)

## Riepilogo finale

- Un **sottogruppo** si verifica con il criterio $a*b^{-1}\in H$: più rapido di controllare tutti gli assiomi separatamente.
- Le **classi laterali** $aH$ formano una partizione di $G$ (sono le classi di equivalenza di $x\sim y\iff x^{-1}y\in H$), e ognuna ha la stessa cardinalità di $H$.
- Il **teorema di Lagrange** ($|H|$ divide $|G|$ per $G$ finito) segue direttamente da questo conteggio, ed è uno strumento aritmetico potentissimo.
- Conseguenza immediata: **ogni gruppo di ordine primo è ciclico** — una classificazione completa ottenuta con un argomento di poche righe.
