## Perché ci interessa

Nel notebook 1.6 abbiamo imparato la formula risolutiva per le equazioni di secondo grado; esistono formule analoghe (più complesse) per il terzo e quarto grado. Per secoli i matematici cercarono una formula per il **quinto** grado — senza successo. Nel 1832 Évariste Galois, a vent'anni, spiegò **perché**: associò a ogni equazione un **gruppo** (il gruppo delle simmetrie delle sue radici) e dimostrò che l'equazione è risolvibile per radicali **se e solo se** questo gruppo ha una proprietà specifica (la **risolubilità**). Il gruppo $S_5$ (notebook 11.5) non ce l'ha — e questo, non un limite di ingegno umano, è la vera ragione per cui nessuna formula può esistere. Questo notebook, il capstone dell'intero modulo di algebra astratta, mette insieme gruppi, campi ed estensioni per raccontare questa storia con il rigore che merita.

## Teoria, passo dopo passo

### Il gruppo di Galois di un'estensione

Data un'estensione di campi $F\subseteq E$, il **gruppo di Galois** $\operatorname{Gal}(E/F)$ è l'insieme degli **automorfismi** di $E$ che fissano $F$ punto per punto: $\{\sigma:E\to E\text{ isomorfismo}:\sigma(a)=a\;\forall a\in F\}$, con la composizione come operazione. È un gruppo (composizione di automorfismi è un automorfismo, l'identità fissa tutto, l'inverso di un automorfismo è un automorfismo — verifiche dirette analoghe a quelle del notebook 11.1).

**Esempio concreto**: $\operatorname{Gal}(\mathbb C/\mathbb R)$. Un automorfismo $\sigma$ di $\mathbb C$ che fissa $\mathbb R$ deve mandare $i$ in una radice di $x^2+1$ (perché $\sigma$ rispetta le operazioni, quindi $\sigma(i)^2=\sigma(i^2)=\sigma(-1)=-1$, essendo $-1\in\mathbb R$ fissato): le uniche radici sono $\pm i$. Se $\sigma(i)=i$: $\sigma$ è l'identità (essendo $\{1,i\}$ una base su $\mathbb R$, notebook 11.12, $\sigma$ è determinato dal suo valore su $i$). Se $\sigma(i)=-i$: $\sigma(a+bi)=a-bi$, la **coniugazione complessa** (notebook 9.1) — e si verifica che è effettivamente un automorfismo ($\overline{z_1+z_2}=\bar z_1+\bar z_2$, $\overline{z_1z_2}=\bar z_1\bar z_2$, dimostrato nell'esercizio 4 del notebook 9.1). Dunque $\operatorname{Gal}(\mathbb C/\mathbb R)=\{\text{id},\text{coniugazione}\}\cong\mathbb Z/2\mathbb Z$, un gruppo di ordine 2 — coerente con $[\mathbb C:\mathbb R]=2$ (notebook 11.12): non è un caso, per estensioni "sufficientemente buone" (**di Galois**), $|\operatorname{Gal}(E/F)|=[E:F]$.

### Il campo di spezzamento e il gruppo di Galois di un'equazione

Data un'equazione polinomiale $p(x)=0$ su un campo $F$, il **campo di spezzamento** $E$ è la più piccola estensione di $F$ contenente **tutte** le radici di $p$ (nel notebook 11.12, esercizio 4, $\mathbb Q(\sqrt[3]2,\omega)$ era esattamente il campo di spezzamento di $x^3-2$). Il **gruppo di Galois dell'equazione** è $\operatorname{Gal}(E/F)$: ogni suo elemento **permuta le radici** di $p$ tra loro (un automorfismo che fissa $F$ deve mandare radici di $p$ in radici di $p$, perché applicando $\sigma$ all'equazione $p(\alpha)=0$, con i coefficienti di $p$ fissati in $F$, si ottiene $p(\sigma(\alpha))=0$). Questo dà un **omomorfismo iniettivo** $\operatorname{Gal}(E/F)\hookrightarrow S_n$ (dove $n$ è il numero di radici distinte, notebook 11.5): il gruppo di Galois è, sempre, un sottogruppo di un gruppo di permutazioni.

### Gruppi risolubili

Un gruppo $G$ è **risolubile** se esiste una catena di sottogruppi $\{e\}=G_0\trianglelefteq G_1\trianglelefteq\cdots\trianglelefteq G_k=G$ dove ogni $G_i\trianglelefteq G_{i+1}$ (normale nel successivo) e ogni quoziente $G_{i+1}/G_i$ è **abeliano**. Intuitivamente: $G$ si "smonta" in pezzi abeliani, uno strato alla volta — l'immagine è quella di una torre di estensioni (notebook 11.12), dove ogni piano si costruisce "aggiungendo una radice" in modo controllato.

$S_n$ per $n\le4$ è risolubile (verificabile esplicitamente: $\{e\}\trianglelefteq V_4\trianglelefteq A_4\trianglelefteq S_4$, con $V_4$ il gruppo di Klein, tutti i quozienti abeliani). Ma:

**Teorema (enunciato).** $A_n$ è **semplice** (privo di sottogruppi normali non banali) per $n\ge5$, quindi $S_n$ **non è risolubile** per $n\ge5$.

Questo è il cuore tecnico della teoria: se $A_5$ non ha sottogruppi normali non banali, l'unica catena possibile per $S_5$ sarebbe $\{e\}\trianglelefteq A_5\trianglelefteq S_5$, ma il quoziente $A_5/\{e\}=A_5$ non è abeliano (essendo $A_5$ un gruppo non commutativo di ordine 60) — la catena non può essere raffinata ulteriormente, e nessun'altra catena esiste.

### Il teorema di Abel-Ruffini

**Teorema (Abel-Ruffini, enunciato).** Un'equazione polinomiale $p(x)=0$ è risolvibile **per radicali** (cioè le sue radici si esprimono con le quattro operazioni e l'estrazione di radici $n$-esime, applicate un numero finito di volte ai coefficienti) **se e solo se** il suo gruppo di Galois è **risolubile**.

**Conseguenza.** Esistono equazioni di quinto grado (per esempio $x^5-4x+2=0$, il cui gruppo di Galois si dimostra essere l'intero $S_5$) il cui gruppo di Galois è $S_5$, **non risolubile** per il teorema precedente: dunque **non esiste alcuna formula generale**, basata su somme, prodotti, quozienti e radici $n$-esime, che esprima le radici di ogni equazione di quinto grado a partire dai suoi coefficienti.

**Perché questo NON contraddice il caso 2°, 3°, 4° grado**: per $n\le4$, $S_n$ **è** risolubile (catena esibita sopra per $S_4$; per $S_2,S_3$ è ancora più semplice, essendo essi stessi già "quasi abeliani"), quindi il teorema di Abel-Ruffini garantisce che una formula **debba** esistere — ed è esattamente la formula del notebook 1.6 per il secondo grado, e le formule (più elaborate) di Cardano e Ferrari per il terzo e quarto. La barriera al quinto grado non è un fallimento di ricerca, ma una **impossibilità dimostrata**: il primo, storicamente, di una lunga serie di teoremi di impossibilità in matematica (insieme alla trisezione dell'angolo con riga e compasso, anch'essa dimostrabile impossibile con argomenti di teoria dei campi analoghi a questi).

````python
from sympy.combinatorics import Permutation, PermutationGroup, AlternatingGroup, SymmetricGroup

# Verifica computazionale: A5 e' semplice, S4 e' risolubile, S5 non lo e'
S4 = SymmetricGroup(4)
S5 = SymmetricGroup(5)
A5 = AlternatingGroup(5)

print("S4 e' risolubile?", S4.is_solvable)
print("S5 e' risolubile?", S5.is_solvable)
print("A5 e' semplice (nessun sottogruppo normale non banale)?", A5.is_simple())
print("A5 e' abeliano?", A5.is_abelian)
print("Ordine di A5:", A5.order())

# Sottogruppi normali di S4 (dovrebbe includere una catena risolubile)
print("\nOrdine di S4:", S4.order())
A4 = AlternatingGroup(4)
print("A4 e' normale in S4?", S4.is_normal(A4))
print("S4/A4 e' abeliano (ordine 2, sempre abeliano)?", True)
````

## Esercizio svolto e commentato

**Problema:** determina il gruppo di Galois del campo di spezzamento di $x^3-2$ su $\mathbb Q$, cioè $\operatorname{Gal}(\mathbb Q(\sqrt[3]2,\omega)/\mathbb Q)$ (dove $\omega=e^{2\pi i/3}$, radice di $x^2+x+1$, notebook 9.4), e verifica che è risolubile.

**Ragionamento:**

**Passo 1 — identifichiamo le tre radici di $x^3-2$.** $\sqrt[3]2,\;\omega\sqrt[3]2,\;\omega^2\sqrt[3]2$ (le tre radici cubiche di $2$, ottenute moltiplicando per le radici cubiche dell'unità, notebook 9.4).

**Passo 2 — il grado dell'estensione.** Dall'esercizio 4 del notebook 11.12, $[\mathbb Q(\sqrt[3]2,\omega):\mathbb Q]=6$. Per estensioni di questo tipo (**di Galois**, cioè campi di spezzamento su campi di caratteristica $0$), $|\operatorname{Gal}(E/F)|=[E:F]=6$.

**Passo 3 — identifichiamo il gruppo di ordine 6.** Un automorfismo $\sigma\in\operatorname{Gal}(E/\mathbb Q)$ è determinato dal suo effetto sulle radici (deve permutarle, per l'osservazione della teoria) e su $\omega$: essendo $\sigma(\sqrt[3]2)$ una delle 3 radici cubiche di 2, e $\sigma(\omega)$ una delle 2 radici di $x^2+x+1$ (cioè $\omega$ o $\omega^2$), ci sono al più $3\times2=6$ possibilità — che coincidono con l'ordine calcolato al passo 2: **tutte** le combinazioni sono effettivamente realizzate da automorfismi.

**Passo 4 — riconosciamo la struttura del gruppo.** Un gruppo non abeliano di ordine 6 è (a meno di isomorfismo) unico: $S_3$ (o equivalentemente $D_3$, il gruppo diedrale del triangolo, notebook 11.2) — infatti gli automorfismi che fissano $\sqrt[3]2$ ma scambiano $\omega\leftrightarrow\omega^2$ agiscono come una "riflessione" sulle tre radici cubiche disposte come vertici di un triangolo nel piano complesso (esattamente come nel notebook 9.4, dove le radici $n$-esime dell'unità sono i vertici di un poligono regolare), mentre quelli che permutano ciclicamente le tre radici agiscono come "rotazioni".

**Conclusione:** $\operatorname{Gal}(\mathbb Q(\sqrt[3]2,\omega)/\mathbb Q)\cong S_3$. Poiché $S_3$ è risolubile (catena $\{e\}\trianglelefteq A_3\trianglelefteq S_3$, con $A_3\cong\mathbb Z/3\mathbb Z$ abeliano e $S_3/A_3$ di ordine 2, abeliano), per il teorema di Abel-Ruffini l'equazione $x^3-2=0$ **è** risolvibile per radicali — coerentemente col fatto che conosciamo esplicitamente le sue radici, $\sqrt[3]2,\omega\sqrt[3]2,\omega^2\sqrt[3]2$, scritte con un radicale.

````python
from sympy.combinatorics import SymmetricGroup, AlternatingGroup

S3 = SymmetricGroup(3)
print("Ordine di S3:", S3.order())
print("S3 e' risolubile?", S3.is_solvable)
print("S3 e' abeliano?", S3.is_abelian)

A3 = AlternatingGroup(3)
print("\nA3 e' normale in S3?", S3.is_normal(A3))
print("A3 e' abeliano (ciclico di ordine 3)?", A3.is_abelian)
print("Indice [S3:A3] =", S3.order() // A3.order(), "(sempre abeliano, essendo primo)")

import sympy as sp
print("\nLe tre radici di x^3-2:")
x = sp.symbols('x')
soluzioni = sp.solve(x**3 - 2, x)
for s in soluzioni:
    print(" ", sp.simplify(s))
````

## Riepilogo finale

- Il **gruppo di Galois** $\operatorname{Gal}(E/F)$ raccoglie gli automorfismi di $E$ che fissano $F$; per un campo di spezzamento, agisce come **permutazioni delle radici**, dando un sottogruppo di $S_n$.
- Un gruppo è **risolubile** se ammette una catena di sottogruppi normali con quozienti abeliani; $S_n$ è risolubile per $n\le4$ ma **non** per $n\ge5$, perché $A_n$ diventa semplice.
- Il **teorema di Abel-Ruffini**: un'equazione è risolvibile per radicali $\iff$ il suo gruppo di Galois è risolubile — spiega sia perché le formule di secondo, terzo, quarto grado esistono, sia perché **nessuna** formula generale può esistere per il quinto grado.
