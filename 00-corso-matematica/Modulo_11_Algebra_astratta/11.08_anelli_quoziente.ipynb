## Teoria, passo dopo passo

### Costruzione dell'anello quoziente

Dato un ideale $I\trianglelefteq R$, l'insieme delle classi laterali $R/I=\{a+I:a\in R\}$ (rispetto al gruppo additivo, notebook 11.3 — sempre ben definito perché $(R,+)$ è abeliano, quindi ogni sottogruppo è normale) diventa un anello con le operazioni:

$$(a+I)+(b+I)=(a+b)+I \qquad\qquad (a+I)\times(b+I)=(a\times b)+I$$

La somma è ben definita per il notebook 11.3 (gruppo abeliano). Verifichiamo che anche il **prodotto** sia ben definito — qui entra in gioco, per la prima volta, la proprietà di **assorbimento** dell'ideale (non bastava per i sottogruppi generici, notebook 11.3).

**Teorema.** Il prodotto su $R/I$ è ben definito.

**Dimostrazione.** Supponiamo $a+I=a'+I$ e $b+I=b'+I$, cioè $a-a'=i_1\in I$ e $b-b'=i_2\in I$. Vogliamo $ab+I=a'b'+I$, cioè $ab-a'b'\in I$. Scriviamo $a=a'+i_1$, $b=b'+i_2$: $ab=(a'+i_1)(b'+i_2)=a'b'+a'i_2+i_1b'+i_1i_2$. Dunque $ab-a'b'=a'i_2+i_1b'+i_1i_2$. Ciascun termine è in $I$: $a'i_2\in I$ (assorbimento, con $a'\in R$ e $i_2\in I$), $i_1b'\in I$ (assorbimento), $i_1i_2\in I$ (assorbimento, o anche solo chiusura). La somma di elementi di $I$ è in $I$ (sottogruppo): $ab-a'b'\in I$. $\blacksquare$

**Perché serve l'assorbimento e non basta la chiusura di un sottoanello**: nei termini $a'i_2$ e $i_1b'$, i fattori $a'$ e $b'$ sono elementi **generici** di $R$, non necessariamente di $I$ — è proprio l'assorbimento (valido per **ogni** $r\in R$, non solo per $r\in I$) a garantire che questi prodotti restino in $I$. Con un semplice sottoanello, questa garanzia mancherebbe, ed è esattamente il fenomeno patologico osservato con $\mathbb Q/\mathbb Z$ nel notebook 11.7.

**Teorema.** $(R/I,+,\times)$ è un anello (commutativo se $R$ lo è, unitario con neutro $1+I$ se $R$ lo è).

La dimostrazione è una verifica diretta degli assiomi, ereditati da $R$ tramite le classi (associatività, distributiva: tutte si riducono a proprietà già vere in $R$, applicate ai rappresentanti — lo stesso schema usato nell'esercizio del notebook 11.1 per $\mathbb Z/n\mathbb Z$, ora generalizzato).

### $\mathbb Z/n\mathbb Z$ come $\mathbb Z/(n)$

Con questa costruzione generale, $\mathbb Z/n\mathbb Z$ (visto fin dal notebook 11.1 come gruppo, poi come anello nel notebook 11.6) è semplicemente $\mathbb Z/(n)$, il quoziente di $\mathbb Z$ per l'ideale principale $(n)$: ora sappiamo **perché** la moltiplicazione modulo $n$ è ben definita — è un caso particolare del teorema appena dimostrato.

### Il primo teorema di isomorfismo per anelli

Un **omomorfismo di anelli** $f:R\to R'$ soddisfa $f(a+b)=f(a)+f(b)$ e $f(a\times b)=f(a)\times f(b)$ (rispetta entrambe le operazioni). Il suo **nucleo** $\ker f=\{a:f(a)=0\}$ è sempre un **ideale** (non solo un sottogruppo): dimostrazione analoga a quella del notebook 11.4 per il nucleo di un omomorfismo di gruppi, con l'assorbimento che segue da $f(ra)=f(r)f(a)=f(r)\times0=0$ per $a\in\ker f$ e $r\in R$ arbitrario.

**Teorema.** $R/\ker f\cong\operatorname{Im}f$ (come anelli).

La dimostrazione ricalca esattamente quella del notebook 11.4 (definire $\varphi(a+\ker f)=f(a)$, verificare buona posizione, iniettività, suriettività su $\operatorname{Im}f$), con la verifica aggiuntiva che $\varphi$ rispetti anche il prodotto: $\varphi((a+\ker f)\times(b+\ker f))=\varphi(ab+\ker f)=f(ab)=f(a)f(b)=\varphi(a+\ker f)\times\varphi(b+\ker f)$.

### Costruire $\mathbb C$ da zero: $\mathbb R[x]/(x^2+1)$

Consideriamo l'omomorfismo di valutazione $\operatorname{ev}_i:\mathbb R[x]\to\mathbb C$, $\operatorname{ev}_i(p(x))=p(i)$ (sostituire $x$ con l'unità immaginaria del notebook 9.1). È suriettivo: ogni $a+bi\in\mathbb C$ è raggiunto dal polinomio $a+bx$. Il nucleo è $\ker(\operatorname{ev}_i)=\{p(x):p(i)=0\}$: si dimostra (usando la divisione euclidea tra polinomi, che vedremo nel notebook 11.11) che $\ker(\operatorname{ev}_i)=(x^2+1)$, l'ideale generato dal polinomio che ha $i$ come radice. Per il primo teorema di isomorfismo:

$$\mathbb R[x]/(x^2+1)\cong\mathbb C$$

**Il significato di questa costruzione**: $\mathbb C$ non è "inventato" aggiungendo un simbolo misterioso $i$ con $i^2=-1$ (come presentato nel notebook 9.1, in modo intuitivo): è, con piena legittimità algebrica, l'anello dei polinomi reali **modulo la relazione** $x^2=-1$ — ogni classe di $\mathbb R[x]/(x^2+1)$ ha un unico rappresentante della forma $a+bx$ (dividendo per $x^2+1$ con resto, esattamente come si riducono le classi resto modulo $n$ in $\mathbb Z$), e la classe di $x$ gioca esattamente il ruolo di $i$.

````python
import sympy as sp

x = sp.symbols('x')

# Costruzione di C come R[x]/(x^2+1): riduciamo alcuni polinomi modulo x^2+1
polinomi = [x**2, x**3, x**3 + 2*x**2 + x + 1, x**4]

modulo = x**2 + 1
for p in polinomi:
    quoziente, resto = sp.div(p, modulo, x)
    print(f"{p} = ({quoziente}) * (x^2+1) + ({resto})  ->  classe: {resto}")

print("\nSostituendo x=i (unita' immaginaria), il resto x^2 si comporta come -1:")
print("x^2 mod (x^2+1) =", sp.div(x**2, modulo, x)[1], " <-> i^2 = -1 in C")
````

## Esercizio svolto e commentato

**Problema:** dimostra che $\mathbb Z[x]/(x)\cong\mathbb Z$ (un altro esempio istruttivo di anello quoziente, più semplice di $\mathbb C$).

**Ragionamento:**

**Passo 1 — definiamo l'omomorfismo di valutazione.** $\operatorname{ev}_0:\mathbb Z[x]\to\mathbb Z$, $\operatorname{ev}_0(p(x))=p(0)$ (valutare il polinomio in $0$, cioè estrarne il termine noto). Verifichiamo che sia un omomorfismo di anelli: $\operatorname{ev}_0(p+q)=(p+q)(0)=p(0)+q(0)=\operatorname{ev}_0(p)+\operatorname{ev}_0(q)$; $\operatorname{ev}_0(pq)=(pq)(0)=p(0)q(0)=\operatorname{ev}_0(p)\operatorname{ev}_0(q)$ (la valutazione in un punto rispetta somma e prodotto di polinomi, una proprietà elementare del calcolo con polinomi).

**Passo 2 — calcoliamo il nucleo.** $\ker(\operatorname{ev}_0)=\{p(x):p(0)=0\}$ — esattamente l'ideale $I$ dell'esercizio 4 del notebook 11.7, dove abbiamo già dimostrato $I=(x)$.

**Passo 3 — calcoliamo l'immagine.** $\operatorname{ev}_0$ è suriettiva su $\mathbb Z$: ogni intero $n$ è raggiunto dal polinomio costante $p(x)=n$, con $\operatorname{ev}_0(n)=n$.

**Conclusione — applichiamo il primo teorema di isomorfismo.** $\mathbb Z[x]/(x)=\mathbb Z[x]/\ker(\operatorname{ev}_0)\cong\operatorname{Im}(\operatorname{ev}_0)=\mathbb Z$. **L'intuizione**: nel quoziente per $(x)$, ogni polinomio è "ridotto" al suo termine noto (perché ogni multiplo di $x$ diventa $0$ nella classe) — la classe di un polinomio $a_0+a_1x+a_2x^2+\cdots$ è semplicemente $[a_0]$, e l'isomorfismo con $\mathbb Z$ formalizza esattamente questa intuizione.

````python
import sympy as sp

x = sp.symbols('x')

def ev0(p):
    return p.subs(x, 0)

# Verifica che ev0 e' un omomorfismo di anelli
p = x**3 + 2*x**2 + 5
q = 3*x**2 - x + 7

print("ev0(p+q) =", ev0(p+q))
print("ev0(p)+ev0(q) =", ev0(p)+ev0(q))
print("ev0(p*q) =", ev0(sp.expand(p*q)))
print("ev0(p)*ev0(q) =", ev0(p)*ev0(q))

# Il nucleo: polinomi con termine noto 0
print("\nEsempi nel nucleo (termine noto 0):", ev0(x**2 - 3*x), ev0(5*x**3 + x))
````

## Riepilogo finale

- L'**anello quoziente** $R/I$ richiede che $I$ sia un **ideale** (non basta un sottoanello): l'assorbimento è precisamente ciò che rende il prodotto tra classi ben definito.
- Il **primo teorema di isomorfismo per anelli** ($R/\ker f\cong\operatorname{Im}f$) ricalca esattamente quello per gruppi.
- $\mathbb C\cong\mathbb R[x]/(x^2+1)$: i numeri complessi, presentati nel notebook 9.1 tramite l'assioma $i^2=-1$, sono in realtà i polinomi reali **modulo la relazione** $x^2=-1$ — una costruzione pienamente rigorosa, senza bisogno di "inventare" nulla.
- Lo stesso schema costruisce $\mathbb Z[\sqrt2]$, $\mathbb R[x]/(x-a)\cong\mathbb R$, e in generale ogni estensione algebrica, anticipando il notebook 11.12.
