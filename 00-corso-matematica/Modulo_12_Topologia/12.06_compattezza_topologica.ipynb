## Perché ci interessa

Nel notebook 10.11 abbiamo definito la compattezza in $\mathbb R$ tramite successioni (ogni successione ha una sottosuccessione convergente) e tramite ricoprimenti (Heine-Borel: chiuso e limitato $\iff$ ogni ricoprimento con aperti ammette un sottoricoprimento finito), usando in modo essenziale la struttura metrica di $\mathbb R$. La definizione tramite **ricoprimenti** non usa affatto la distanza — solo aperti — e si generalizza perfettamente a qualunque spazio topologico. Questo notebook riformula la compattezza in questi termini generali, e ne mostra la conseguenza più importante: il teorema di Weierstrass (una funzione continua su un compatto ha massimo e minimo) come caso particolare di un fatto puramente topologico.

## Teoria, passo dopo passo

### Ricoprimenti aperti

Un **ricoprimento aperto** di $X$ è una famiglia $\{U_i\}_{i\in I}$ di aperti la cui unione è $X$ (ogni punto è coperto da almeno un $U_i$). Un **sottoricoprimento** è una sottofamiglia $\{U_i\}_{i\in J}$, $J\subseteq I$, che ancora ricopre $X$.

### Definizione di spazio compatto

$X$ è **compatto** se ogni ricoprimento aperto di $X$ ammette un sottoricoprimento **finito**.

**Perché $(0,1)$ non è compatto:** il ricoprimento $\{(1/n,1):n\ge2\}$ copre $(0,1)$ (ogni $x\in(0,1)$ sta in $(1/n,1)$ per $n$ abbastanza grande, precisamente $n>1/x$), ma nessuna sottofamiglia **finita** lo copre: presi finiti indici $n_1<\cdots<n_k$, l'unione $(1/n_1,1)\cup\cdots\cup(1/n_k,1)=(1/n_k,1)$ (il più grande intervallo, avendo $n_k$ il denominatore più grande, quindi $1/n_k$ il più piccolo) non copre i punti $x\le1/n_k$, come $x=1/(2n_k)$. Questo controesempio ricorda da vicino la dimostrazione, nel notebook 10.11, che $(0,1)$ non è compatto perché non è chiuso (esiste una successione, come $1/n$, che "vorrebbe convergere" a $0\notin(0,1)$): qui vediamo lo stesso fenomeno dal lato dei ricoprimenti.

### Compattezza e chiusura: sottospazi compatti sono chiusi (in spazi di Hausdorff)

**Teorema.** Se $X$ è uno spazio di **Hausdorff** (cioè: per ogni coppia di punti distinti $x\ne y$ esistono aperti disgiunti $U\ni x$, $V\ni y$ — una proprietà di "separazione" che $\mathbb R^n$ con la topologia euclidea soddisfa sempre) e $K\subseteq X$ è compatto, allora $K$ è chiuso in $X$.

**Dimostrazione.** Mostriamo che $X\setminus K$ è aperto. Sia $y\notin K$. Per ogni $x\in K$ (essendo $x\ne y$), per Hausdorff esistono aperti disgiunti $U_x\ni x$, $V_x\ni y$. La famiglia $\{U_x\}_{x\in K}$ è un ricoprimento aperto di $K$; per compattezza, esiste un sottoricoprimento finito $U_{x_1},\ldots,U_{x_n}$. Sia $V=V_{x_1}\cap\cdots\cap V_{x_n}$ (intersezione **finita** di aperti, quindi aperto, notebook 12.1): $V$ è un intorno di $y$ disgiunto da ogni $U_{x_i}$, quindi disgiunto dalla loro unione, che contiene $K$. Dunque $V\subseteq X\setminus K$: ogni $y\notin K$ ha un intorno interamente fuori da $K$, cioè $X\setminus K$ è aperto, cioè $K$ è chiuso. $\blacksquare$ (Il ruolo cruciale della finitezza: senza il passaggio al sottoricoprimento *finito*, l'intersezione $\bigcap_{x\in K}V_x$ potrebbe non essere aperta, notebook 12.1.)

### Continuità e compattezza: generalizzazione del teorema di Weierstrass

**Teorema.** Se $f:X\to Y$ è continua e $X$ è compatto, allora $f(X)$ è compatto.

**Dimostrazione.** Sia $\{V_i\}_{i\in I}$ un ricoprimento aperto di $f(X)$. Ogni $f^{-1}(V_i)$ è aperto in $X$ (per continuità, notebook 12.4), e $\{f^{-1}(V_i)\}_{i\in I}$ ricopre $X$ (ogni $x\in X$ ha $f(x)\in f(X)$, quindi $f(x)\in V_i$ per qualche $i$, quindi $x\in f^{-1}(V_i)$). Per compattezza di $X$, esiste un sottoricoprimento finito $f^{-1}(V_{i_1}),\ldots,f^{-1}(V_{i_n})$. Allora $\{V_{i_1},\ldots,V_{i_n}\}$ ricopre $f(X)$: se $y=f(x)\in f(X)$, $x$ sta in qualche $f^{-1}(V_{i_k})$, quindi $y=f(x)\in V_{i_k}$. Dunque $f(X)$ ammette un sottoricoprimento finito da ogni ricoprimento: $f(X)$ è compatto. $\blacksquare$

**Corollario (teorema di Weierstrass, versione topologica).** Se $f:X\to\mathbb R$ è continua e $X$ è compatto, allora $f(X)$ è un compatto di $\mathbb R$, quindi (per Heine-Borel, notebook 10.11) chiuso e limitato — un insieme chiuso e limitato di $\mathbb R$ ammette sempre massimo e minimo (essendo limitato ha $\sup$ e $\inf$ per completezza, notebook 10.9, ed essendo chiuso li contiene, per il notebook 12.3). Dunque $f$ ha massimo e minimo assoluti — esattamente il teorema dimostrato in $\mathbb R$ nel notebook 10.12, ora visto come istanza di un fenomeno che vale in ogni spazio topologico compatto.

```python
import matplotlib.pyplot as plt
import numpy as np

# Visualizziamo perche' (0,1) non e' compatto: un ricoprimento senza sottoricoprimento finito
fig, ax = plt.subplots(figsize=(8, 4))

n_valori = [2, 3, 4, 6, 10, 20, 50]
for i, n in enumerate(n_valori):
    ax.plot([1/n, 1], [i, i], linewidth=3, label=f"(1/{n}, 1)")

ax.set_yticks(range(len(n_valori)))
ax.set_yticklabels([f"n={n}" for n in n_valori])
ax.set_xlabel("x")
ax.set_title("Ricoprimento {(1/n,1)} di (0,1): nessuna sottofamiglia finita basta")
ax.legend(loc='center right', fontsize=8)
plt.tight_layout()
plt.show()

# Verifica: qualunque sottofamiglia finita di indici lascia scoperta una zona vicino a 0
indici_scelti = [2, 3, 4]  # una sottofamiglia finita a caso
n_max = max(indici_scelti)
punto_non_coperto = 1 / (2 * n_max)
print(f"Sottofamiglia finita scelta: n in {indici_scelti}")
print(f"Il punto x={punto_non_coperto} non e' coperto da nessuno di questi intervalli (verifica: x < 1/{n_max})")
```

## Esercizio svolto e commentato

**Problema:** dimostra che ogni sottoinsieme **finito** di uno spazio topologico $X$ è compatto, e che, in uno spazio di Hausdorff, ogni sottoinsieme finito è anche chiuso.

**Ragionamento:**

**Passo 1 — compattezza di un insieme finito.** Sia $F=\{x_1,\ldots,x_n\}\subseteq X$, e $\{U_i\}_{i\in I}$ un ricoprimento aperto di $F$. Per ogni $k=1,\ldots,n$, essendo $x_k\in F\subseteq\bigcup_i U_i$, esiste (almeno) un indice $i_k$ con $x_k\in U_{i_k}$. La sottofamiglia finita $\{U_{i_1},\ldots,U_{i_n}\}$ (al più $n$ insiemi, alcuni eventualmente ripetuti) copre $F$, perché ogni $x_k$ sta nel corrispondente $U_{i_k}$. Questo è un sottoricoprimento finito: $F$ è compatto.

**Passo 2 — chiusura in uno spazio di Hausdorff.** Per il teorema della teoria (compatti in Hausdorff sono chiusi), essendo $F$ compatto (passo 1) e $X$ di Hausdorff per ipotesi, $F$ è chiuso.

**Osservazione conclusiva:** questo esercizio spiega perché, in $\mathbb R^n$ (che è di Hausdorff — due punti distinti hanno sempre palle disgiunte attorno, prendendo raggio minore della metà della loro distanza), ogni insieme finito di punti è automaticamente chiuso: fatto ovvio "a occhio" ma che ora discende da un principio generale, non da un calcolo diretto con le distanze. **Attenzione**: la compattezza di per sé (passo 1) **non** richiede Hausdorff — vale in ogni spazio topologico, incluso quello indiscreto, dove però un insieme finito non è affatto detto sia chiuso (nella topologia indiscreta, gli unici chiusi sono $\emptyset$ e $X$).

```python
# Verifica computazionale: ogni sottoinsieme finito di R^n e' chiuso (esempio concreto)
import numpy as np

punti = np.array([[0, 0], [1, 1], [2, -1]])  # F = 3 punti in R^2

def distanza_minima_tra_punti(punti):
    n = len(punti)
    min_dist = np.inf
    for i in range(n):
        for j in range(i+1, n):
            d = np.linalg.norm(punti[i] - punti[j])
            min_dist = min(min_dist, d)
    return min_dist

d_min = distanza_minima_tra_punti(punti)
print(f"Distanza minima tra i punti di F: {d_min:.4f}")
print(f"Raggio sicuro per palle disgiunte attorno a ciascun punto: {d_min/2:.4f}")

# Ogni punto y fuori da F ha un intorno che non tocca F (dimostra che il complementare e' aperto)
y_test = np.array([0.5, 0.5])
distanze_da_F = [np.linalg.norm(y_test - p) for p in punti]
raggio_sicuro = min(distanze_da_F)
print(f"\nPunto y={y_test}, fuori da F. Distanza minima da F: {raggio_sicuro:.4f}")
print(f"Una palla di raggio {raggio_sicuro:.4f} attorno a y non tocca F: complementare di F aperto in quel punto")
```

**4.** Per ogni $y\in K_2$: essendo $y\notin K_1$ (disgiunti), applicando l'argomento della dimostrazione del teorema della teoria (con $K=K_1$ e il punto $y$), esistono aperti disgiunti $U_y\supseteq K_1$ e $V_y\ni y$. La famiglia $\{V_y\}_{y\in K_2}$ ricopre $K_2$; per compattezza di $K_2$, esiste un sottoricoprimento finito $V_{y_1},\ldots,V_{y_n}$. ## Riepilogo finale

- Uno spazio è **compatto** se ogni ricoprimento con aperti ammette un sottoricoprimento **finito** — la formulazione topologica pura della compattezza, senza riferimento a distanze o successioni.
- Nei spazi di **Hausdorff** (dove punti distinti hanno intorni disgiunti — vero per $\mathbb R^n$), ogni compatto è chiuso: la finitezza del sottoricoprimento è ciò che rende l'intersezione degli intorni "buoni" ancora aperta.
- La compattezza è preservata da funzioni continue ($f$ continua, $X$ compatto $\Rightarrow$ $f(X)$ compatto), da cui segue il teorema di Weierstrass come caso particolare in $\mathbb R$.
- Chiusi dentro compatti sono compatti; unioni finite di compatti sono compatte; in spazi di Hausdorff, compatti disgiunti si possono sempre separare con aperti disgiunti.
