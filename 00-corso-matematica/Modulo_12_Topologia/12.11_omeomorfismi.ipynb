## Perché ci interessa

Nel notebook 11.4 un **isomorfismo** di gruppi era un omomorfismo biettivo con inverso anch'esso omomorfismo: due gruppi isomorfi sono "algebricamente identici", anche se i loro elementi sono oggetti diversi. In topologia, la nozione analoga è quella di **omeomorfismo**: due spazi omeomorfi sono "topologicamente identici" — hanno esattamente gli stessi aperti, a meno di rietichettare i punti. Questo notebook definisce con precisione questa equivalenza e fornisce gli strumenti (gli **invarianti topologici**, come compattezza e connessione) per dimostrare che due spazi **non** sono omeomorfi, un compito spesso più delicato che dimostrare che lo sono.

## Teoria, passo dopo passo

### Definizione di omeomorfismo

Una funzione $f:X\to Y$ è un **omeomorfismo** se è biettiva, continua, e con inversa $f^{-1}:Y\to X$ anch'essa continua. Se esiste un omeomorfismo tra $X$ e $Y$, i due spazi si dicono **omeomorfi**, scritto $X\cong Y$.

**Attenzione: biettiva e continua non basta.** Consideriamo $f:[0,2\pi)\to S^1$, $f(t)=(\cos t,\sin t)$ (la stessa costruzione del notebook 12.10, ma qui su un intervallo semiaperto invece che sul quoziente). $f$ è biettiva e continua, ma **non** è un omeomorfismo: la sua inversa non è continua nel punto $(1,0)=f(0)$. Intuitivamente, punti su $S^1$ vicini a $(1,0)$ "da sotto" (con angolo vicino a $2\pi$) corrispondono, tramite $f^{-1}$, a valori $t$ vicini a $2\pi$ — **lontani** da $t=0=f^{-1}((1,0))$ nella metrica di $[0,2\pi)$. Questo esempio, dimostrato in dettaglio nell'esercizio svolto, mostra che la richiesta di continuità dell'inversa è **sostanziale**, non una formalità.

### Omeomorfismo come relazione di equivalenza

Analogamente all'isomorfismo di gruppi (notebook 11.4), l'omeomorfismo è una relazione di equivalenza sugli spazi topologici: **riflessiva** ($\operatorname{id}_X$ è sempre un omeomorfismo), **simmetrica** (se $f$ è omeomorfismo, $f^{-1}$ lo è, con inversa $f$), **transitiva** (composizione di omeomorfismi è un omeomorfismo, per la composizione di funzioni continue del notebook 12.4, applicata sia a $g\circ f$ sia a $f^{-1}\circ g^{-1}=(g\circ f)^{-1}$).

### Invarianti topologici

Una proprietà $P$ di uno spazio topologico è un **invariante topologico** se, quando $X\cong Y$, $X$ ha $P$ se e solo se $Y$ ha $P$. Compattezza e connessione sono invarianti: se $f:X\to Y$ è un omeomorfismo (in particolare, continua e suriettiva), per i teoremi dei notebook 12.5 e 12.6, $X$ compatto/connesso $\Rightarrow$ $Y=f(X)$ compatto/connesso; applicando lo stesso argomento a $f^{-1}$ (anch'essa un omeomorfismo), vale anche il viceversa.

**Uso pratico degli invarianti: dimostrare la NON omeomorfia.** Se $X$ ha una proprietà invariante che $Y$ non ha (o viceversa), allora $X\not\cong Y$. Per esempio: $[0,1]$ è compatto e connesso; $(0,1)$ è connesso ma **non** compatto (notebook 12.6, 12.8); dunque $[0,1]\not\cong(0,1)$ — nessuna funzione biettiva bicontinua può esistere tra i due, anche se hanno "la stessa cardinalità" come insiemi (notebook 10.5: entrambi hanno la cardinalità del continuo).

### Teorema: un altro invariante utile, la connessione dopo rimozione di un punto

**Teorema.** Se $f:X\to Y$ è un omeomorfismo e $x_0\in X$, allora $X\setminus\{x_0\}\cong Y\setminus\{f(x_0)\}$ (la restrizione di $f$ è ancora un omeomorfismo tra i due spazi privati di un punto corrispondente).

**Dimostrazione.** $f|_{X\setminus\{x_0\}}:X\setminus\{x_0\}\to Y\setminus\{f(x_0)\}$ è biettiva (restrizione di una biezione a un dominio con un punto tolto, verso il codominio con l'immagine di quel punto tolta), continua (restrizione di funzione continua), e la sua inversa è $f^{-1}|_{Y\setminus\{f(x_0)\}}$, anch'essa continua per lo stesso motivo. $\blacksquare$ Questo teorema, combinato con la connessione, è lo strumento standard per dimostrare che una retta e una circonferenza non sono omeomorfe: **togliendo un punto**, $\mathbb R\setminus\{x_0\}$ diventa sconnesso (due semirette), mentre $S^1$ privato di un punto **resta connesso** (un arco, omeomorfo a un intervallo aperto) — proprietà diverse dopo la stessa operazione, quindi gli spazi originali non potevano essere omeomorfi.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# R privato di un punto: sconnesso (due pezzi)
x1 = np.linspace(-3, -0.05, 50)
x2 = np.linspace(0.05, 3, 50)
axes[0].plot(x1, np.zeros_like(x1), 'b-', linewidth=3)
axes[0].plot(x2, np.zeros_like(x2), 'b-', linewidth=3)
axes[0].plot(0, 0, 'wo', markeredgecolor='red', markersize=10, label='punto rimosso')
axes[0].set_title("R \\ {0}: SCONNESSO (due semirette)")
axes[0].legend()
axes[0].set_yticks([])

# S^1 privato di un punto: connesso (un arco)
theta = np.linspace(0.05, 2*np.pi - 0.05, 200)
axes[1].plot(np.cos(theta), np.sin(theta), 'g-', linewidth=3)
axes[1].plot(1, 0, 'wo', markeredgecolor='red', markersize=10, label='punto rimosso')
axes[1].set_aspect('equal')
axes[1].set_title("S^1 \\ {punto}: CONNESSO (un arco)")
axes[1].legend()

plt.tight_layout()
plt.show()

print("R privato di un punto ha 2 componenti connesse; S^1 privato di un punto ne ha 1.")
print("Essendo la connessione un invariante topologico, R e S^1 NON sono omeomorfi.")
```

## Esercizio svolto e commentato

**Problema:** dimostra in dettaglio che $f:[0,2\pi)\to S^1$, $f(t)=(\cos t,\sin t)$, è biettiva e continua ma **non** è un omeomorfismo (la sua inversa non è continua in $(1,0)$).

**Ragionamento:**

**Passo 1 — biettività.** $f$ è suriettiva (ogni punto di $S^1$ ha un angolo $t\in[0,2\pi)$) e iniettiva (angoli distinti in $[0,2\pi)$ danno punti distinti su $S^1$, essendo $\cos,\sin$ periodiche di periodo $2\pi$ e $[0,2\pi)$ un solo periodo). Continua: dimostrato nell'esercizio svolto del notebook 12.10 (stessa funzione).

**Passo 2 — costruiamo una successione che mette in crisi la continuità dell'inversa.** Consideriamo $y_n=(\cos(2\pi-1/n),\sin(2\pi-1/n))\in S^1$, per $n\ge1$. Al crescere di $n$, l'angolo $2\pi-1/n\to2\pi$, quindi (per continuità di $\cos,\sin$) $y_n\to(\cos2\pi,\sin2\pi)=(1,0)$ in $S^1$.

**Passo 3 — calcoliamo $f^{-1}(y_n)$ e verifichiamo che non converge a $f^{-1}((1,0))$.** $f^{-1}(y_n)=2\pi-1/n$ (l'unico angolo in $[0,2\pi)$ corrispondente). Ma $f^{-1}((1,0))=0$ (l'unico angolo in $[0,2\pi)$ con $\cos=1,\sin=0$ è $t=0$). La successione $f^{-1}(y_n)=2\pi-1/n\to2\pi$, **non** a $0$: se $f^{-1}$ fosse continua, dovrebbe valere $f^{-1}(y_n)\to f^{-1}(\lim y_n)=f^{-1}((1,0))=0$ (continuità e limiti di successioni sono collegati, come nel notebook 10.12), ma $2\pi-1/n\to2\pi\ne0$.

**Conclusione:** $f^{-1}$ non è continua in $(1,0)$: $f$ non è un omeomorfismo. **Il punto concettuale:** $[0,2\pi)$ "non sa" che i suoi due estremi sono topologicamente vicini su $S^1$ — la metrica di $[0,2\pi)$ misura $2\pi-1/n$ come vicino a $2\pi$ (fuori dall'intervallo) e lontano da $0$, mentre su $S^1$ i punti corrispondenti sono vicinissimi. Questo è precisamente il tipo di "incollamento mancante" che lo spazio quoziente $[0,1]/\!\sim$ del notebook 12.10 (con $0\sim1$ esplicitamente identificati) risolve correttamente, a differenza dell'intervallo semiaperto.

```python
import numpy as np

n_vals = np.array([1, 5, 20, 100, 1000])
angoli = 2*np.pi - 1/n_vals

y_n = np.stack([np.cos(angoli), np.sin(angoli)], axis=1)

print("n\tangolo f^-1(y_n)\ty_n (su S^1)")
for n, ang, y in zip(n_vals, angoli, y_n):
    print(f"{n}\t{ang:.5f}\t\t({y[0]:.5f}, {y[1]:.5f})")

print(f"\ny_n converge a (1,0): {np.allclose(y_n[-1], [1,0], atol=1e-2)}")
print(f"f^-1(y_n) converge a 2*pi = {2*np.pi:.5f}, NON a f^-1((1,0))=0")
print("-> f^-1 non e' continua in (1,0): f non e' un omeomorfismo")
```

## Esercizi da risolvere

1. Dimostra che $(0,1)\cong\mathbb R$ (sono omeomorfi), esibendo esplicitamente un omeomorfismo (suggerimento: una funzione trigonometrica o una funzione razionale che manda $(0,1)$ biettivamente su $\mathbb R$, come $\tan$ opportunamente riscalata).
2. Dimostra che $[0,1]\not\cong[0,1)$ usando un invariante topologico opportuno (suggerimento: compattezza).
3. Dimostra che $\mathbb R\not\cong\mathbb R^2$, usando l'invariante "connessione dopo rimozione di un punto" del teorema della teoria (suggerimento: $\mathbb R^2$ privato di un punto resta connesso — è addirittura "connesso per archi", concetto intuitivamente chiaro: si può sempre girare attorno al punto mancante).
4. *(più impegnativo)* Dimostra che $S^1\not\cong[0,1]$ usando due invarianti diversi in combinazione: entrambi sono compatti e connessi (quindi questi due invarianti da soli non bastano a distinguerli), ma la rimozione di **un punto qualsiasi** lascia $S^1$ connesso mentre esiste un punto (l'estremo) la cui rimozione da $[0,1]$ lo lascia connesso, e un altro punto (interno) la cui rimozione lo sconnette — descrivi con precisione questa differenza e perché è sufficiente.

**1.** La funzione $f(x)=\tan\!\big(\pi(x-\tfrac12)\big)$ manda $(0,1)$ in $\mathbb R$: quando $x\to0^+$, $\pi(x-1/2)\to-\pi/2$, e $\tan\to-\infty$; quando $x\to1^-$, $\pi(x-1/2)\to\pi/2$, e $\tan\to+\infty$; $\tan$ è continua, strettamente crescente e biettiva su $(-\pi/2,\pi/2)$ verso $\mathbb R$ (fatto standard di trigonometria). L'inversa $f^{-1}(y)=\frac1\pi\arctan(y)+\frac12$ è anch'essa continua (composizione di $\arctan$, continua su tutto $\mathbb R$, con operazioni algebriche continue). Dunque $f$ è un omeomorfismo: $(0,1)\cong\mathbb R$.

**2.** $[0,1]$ è compatto (chiuso e limitato in $\mathbb R$, per Heine-Borel, notebook 10.11/12.6). $[0,1)$ **non** è compatto: il ricoprimento $\{[0,1-1/n):n\ge2\}$ non ammette sottoricoprimento finito (stesso argomento del notebook 12.6 per $(0,1)$, adattato). Essendo la compattezza un invariante topologico, e $[0,1]$ compatto mentre $[0,1)$ non lo è, i due spazi non sono omeomorfi.

**3.** Supponiamo per assurdo $\mathbb R\cong\mathbb R^2$ tramite un omeomorfismo $f$. Per il teorema della teoria, $\mathbb R\setminus\{x_0\}\cong\mathbb R^2\setminus\{f(x_0)\}$ per ogni $x_0$. Ma $\mathbb R\setminus\{x_0\}$ è sconnesso (due semirette, come nel grafico della teoria), mentre $\mathbb R^2\setminus\{p\}$ (per qualunque punto $p$) è connesso: dati due punti $a,b\ne p$, si può sempre tracciare un cammino continuo da $a$ a $b$ che eviti $p$ (per esempio deviando attorno a $p$ con un piccolo arco), quindi $\mathbb R^2\setminus\{p\}$ è connesso per archi, e ogni spazio connesso per archi è connesso. Essendo la connessione un invariante, $\mathbb R\setminus\{x_0\}\not\cong\mathbb R^2\setminus\{f(x_0)\}$ per nessuna scelta — contraddizione. Dunque $\mathbb R\not\cong\mathbb R^2$.

**4.** Su $S^1$, **ogni** punto rimosso lascia uno spazio connesso (un arco, come nella figura della teoria — per simmetria, non importa quale punto si toglie). Su $[0,1]$, rimuovendo un **estremo** (per esempio $0$) si ottiene $(0,1]$, connesso (è ancora un intervallo, per il notebook 12.5); ma rimuovendo un punto **interno** (per esempio $1/2$) si ottiene $[0,1/2)\cup(1/2,1]$, che **non** è un intervallo, quindi sconnesso (notebook 12.5, esercizio svolto). Se esistesse un omeomorfismo $f:S^1\to[0,1]$, per il teorema della teoria $S^1\setminus\{x_0\}\cong[0,1]\setminus\{f(x_0)\}$ per ogni $x_0$; ma il membro sinistro è **sempre** connesso, mentre il membro destro è sconnesso per la scelta $f(x_0)=1/2$ (cioè $x_0=f^{-1}(1/2)$, un punto che esiste per suriettività) — contraddizione. Dunque $S^1\not\cong[0,1]$: la distinzione emerge non dal confrontare compattezza o connessione degli spazi originali (identiche), ma dal confrontare **come cambiano** dopo la rimozione di un punto, a seconda di quale punto si sceglie.

## Riepilogo finale

- Un **omeomorfismo** è una biezione continua con inversa continua: la nozione precisa di "stessa forma topologica". Biettiva e continua **non basta** — serve anche la continuità dell'inversa, come mostra l'esempio $[0,2\pi)\to S^1$.
- Compattezza e connessione sono **invarianti topologici**: si preservano sotto omeomorfismo, e la loro assenza in uno dei due spazi basta a escludere l'omeomorfia.
- La tecnica "rimuovi un punto e guarda cosa succede alla connessione" è uno strumento potente: distingue $\mathbb R$ da $\mathbb R^2$, e — con un'analisi più fine, punto per punto — $S^1$ da $[0,1]$.
- Dimostrare che due spazi **sono** omeomorfi richiede di esibire un omeomorfismo esplicito; dimostrare che **non lo sono** richiede di trovare un invariante che li distingua — spesso il compito più creativo dei due.
