## Perché ci interessa

Nel notebook 12.1 abbiamo detto che la topologia euclidea su $\mathbb R$ è "l'insieme di tutte le unioni di intervalli aperti": per descriverla non abbiamo elencato ogni singolo aperto (ce ne sono infiniti, e la maggior parte è complicata, come $(0,1)\cup(2,3)\cup(5,7)$), ma abbiamo detto quali sono i "mattoni" — gli intervalli $(a,b)$ — da cui ogni aperto si ottiene per unione. Questa idea, generalizzata, è quella di **base**: una famiglia più piccola e maneggevole di insiemi che *genera* l'intera topologia. Avere una base rende molto più facile sia definire una topologia su un insieme nuovo, sia verificare proprietà (come la continuità, notebook 12.4) controllandole solo sugli elementi della base invece che su tutti gli aperti.

## Teoria, passo dopo passo

### Definizione di base

Sia $(X,\tau)$ uno spazio topologico. Una famiglia $\mathcal B\subseteq\tau$ è una **base** per $\tau$ se ogni aperto $U\in\tau$ si scrive come unione di elementi di $\mathcal B$. Equivalentemente (e più operativamente): $\mathcal B$ è una base se per ogni $U\in\tau$ e ogni $x\in U$ esiste $B\in\mathcal B$ con $x\in B\subseteq U$.

**Esempio:** in $\mathbb R$ con la topologia euclidea, $\mathcal B=\{(a,b):a<b\}$ (tutti gli intervalli aperti) è una base: ogni aperto è per definizione unione di tali intervalli. In $\mathbb R^n$, $\mathcal B=\{B(x,r):x\in\mathbb R^n,r>0\}$ (le palle aperte del notebook 12.1) è una base per la topologia euclidea.

### Da una famiglia a una topologia: quando $\mathcal B$ genera una topologia

Il problema inverso è più utile in pratica: data una famiglia $\mathcal B$ di sottoinsiemi di $X$, quando esiste (ed è unica) una topologia $\tau_{\mathcal B}$ per cui $\mathcal B$ è una base?

**Teorema.** Una famiglia $\mathcal B\subseteq\mathcal P(X)$ è base per **qualche** topologia su $X$ se e solo se: (i) $\bigcup_{B\in\mathcal B}B=X$ (ogni punto è coperto), e (ii) per ogni $B_1,B_2\in\mathcal B$ e ogni $x\in B_1\cap B_2$, esiste $B_3\in\mathcal B$ con $x\in B_3\subseteq B_1\cap B_2$.

**Dimostrazione (schema).** *Necessità*: se $\mathcal B$ è base di una topologia $\tau$, allora $X\in\tau$ dà (i); e $B_1\cap B_2\in\tau$ (intersezione finita di aperti, assioma 3 del notebook 12.1) si scrive come unione di elementi di $\mathcal B$, quindi contiene un $B_3\ni x$ dentro $B_1\cap B_2$, dando (ii). *Sufficienza*: definiamo $\tau_{\mathcal B}=\{$unioni arbitrarie di elementi di $\mathcal B\}\cup\{\emptyset\}$, e verifichiamo i tre assiomi: $\emptyset\in\tau_{\mathcal B}$ per costruzione, $X\in\tau_{\mathcal B}$ per (i); le unioni di elementi di $\tau_{\mathcal B}$ sono ancora unioni di elementi di $\mathcal B$ (unione di unioni), quindi in $\tau_{\mathcal B}$; per le intersezioni finite, la condizione (ii) è esattamente ciò che serve per mostrare che $U_1\cap U_2$ (con $U_1,U_2$ unioni di elementi di $\mathcal B$) è ancora unione di elementi di $\mathcal B$, ragionando punto per punto come nella dimostrazione dell'assioma (3) per gli intervalli in $\mathbb R$ (notebook 12.1). $\blacksquare$

La condizione (ii) è precisamente ciò che fallisce se, per esempio, si prova a usare come base i **quadrati** (invece delle palle) in $\mathbb R^2$ senza includere le loro intersezioni: due quadrati possono intersecarsi in una regione che non è un quadrato, quindi serve un terzo quadrato più piccolo dentro l'intersezione — cosa sempre possibile, ed è per questo che i quadrati (o i rettangoli) *funzionano* come base della topologia euclidea, esattamente come le palle.

### Base numerabile e secondo assioma di numerabilità

Uno spazio si dice a **base numerabile** se ammette una base $\mathcal B$ con $|\mathcal B|=\aleph_0$ (numerabile, notebook 10.5). $\mathbb R^n$ con la topologia euclidea è a base numerabile: basta prendere le palle $B(q,r)$ con $q\in\mathbb Q^n$ (i razionali sono numerabili e densi, notebook 10.5 e 10.9) e $r\in\mathbb Q^+$ — un insieme numerabile per numerabile, quindi numerabile (notebook 10.5), ma sufficiente a generare tutta la topologia, perché ogni palla $B(x,r)$ contiene una palla $B(q,r')$ con centro razionale che contiene ancora $x$.

```python
import matplotlib.pyplot as plt
import numpy as np

# Illustriamo la condizione (ii): l'intersezione di due palle contiene una palla piu' piccola
fig, ax = plt.subplots(figsize=(6, 6))
theta = np.linspace(0, 2*np.pi, 200)

c1, r1 = np.array([0, 0]), 1.3
c2, r2 = np.array([1.2, 0]), 1.3
ax.plot(c1[0] + r1*np.cos(theta), c1[1] + r1*np.sin(theta), label="B(x1, r1)")
ax.plot(c2[0] + r2*np.cos(theta), c2[1] + r2*np.sin(theta), label="B(x2, r2)")

# Un punto x nell'intersezione, e una palla piccola attorno a x contenuta in entrambe
x = np.array([0.6, 0.3])
r3 = 0.35  # scelta abbastanza piccola da stare in entrambe
ax.plot(x[0] + r3*np.cos(theta), x[1] + r3*np.sin(theta), 'r--', label="B(x, r3) dentro l'intersezione")
ax.plot(*x, 'ko')
ax.set_aspect('equal')
ax.legend()
ax.set_title("Condizione (ii): esiste sempre una palla piu' piccola dentro l'intersezione")
plt.tight_layout()
plt.show()

# Verifica numerica: ogni punto dell'intersezione ammette un raggio positivo che ci sta dentro
def dentro_intersezione(p, c1, r1, c2, r2, raggio_test):
    return (np.linalg.norm(p - c1) + raggio_test <= r1) and (np.linalg.norm(p - c2) + raggio_test <= r2)

raggio_max = min(r1 - np.linalg.norm(x - c1), r2 - np.linalg.norm(x - c2))
print(f"Raggio massimo per una palla centrata in x e contenuta nell'intersezione: {raggio_max:.3f}")
print("E' positivo?", raggio_max > 0)
```

## Esercizio svolto e commentato

**Problema:** dimostra che i **rettangoli aperti** $\mathcal B=\{(a,b)\times(c,d):a<b,c<d\}$ formano una base per una topologia su $\mathbb R^2$, e che questa topologia coincide con la topologia euclidea (generata dalle palle).

**Ragionamento:**

**Passo 1 — verifica della condizione (i).** $\mathbb R^2=\bigcup(a,b)\times(c,d)$ (basta prendere, per esempio, un solo rettangolo con $a,c\to-\infty$ e $b,d\to+\infty$, o più correttamente l'unione su tutti gli interi $n$ dei rettangoli $(-n,n)\times(-n,n)$): ogni punto di $\mathbb R^2$ sta in qualche rettangolo. ✓

**Passo 2 — verifica della condizione (ii).** L'intersezione di due rettangoli $(a_1,b_1)\times(c_1,d_1)$ e $(a_2,b_2)\times(c_2,d_2)$ è ancora un rettangolo (aperto), $(\max(a_1,a_2),\min(b_1,b_2))\times(\max(c_1,c_2),\min(d_1,d_2))$ — se non vuoto. Per ogni $x$ nell'intersezione, l'intersezione **stessa** è un elemento di $\mathcal B$ contenente $x$ e contenuta nell'intersezione: la condizione (ii) è verificata con $B_3=B_1\cap B_2$ stesso (un caso particolarmente semplice, più forte del richiesto). ✓ Per il teorema della teoria, $\mathcal B$ è base di una topologia $\tau_{rett}$.

**Passo 3 — $\tau_{rett}=\tau_{euclidea}$: ogni palla contiene un rettangolo e viceversa.** Dato $x\in B(x_0,r)$ (palla), il quadrato aperto centrato in $x_0$ di lato $r/\sqrt2$ (cioè $(x_{0,1}-r/\sqrt2,x_{0,1}+r/\sqrt2)\times(x_{0,2}-r/\sqrt2,x_{0,2}+r/\sqrt2)$) è contenuto in $B(x_0,r)$: ogni suo punto dista da $x_0$ al più $\sqrt{(r/\sqrt2)^2+(r/\sqrt2)^2}=r$ (diagonale del quadrato), quindi sta nella palla chiusa di raggio $r$; con un lato leggermente più piccolo si ottiene la palla aperta. Viceversa, dato un rettangolo $(a,b)\times(c,d)$ e un punto $x$ al suo interno, la palla di raggio $r=\min(x_1-a,b-x_1,x_2-c,d-x_2)$ (la minima distanza di $x$ da uno dei quattro lati) è contenuta nel rettangolo.

**Conclusione:** ogni aperto della topologia euclidea (unione di palle) è anche unione di rettangoli (sostituendo ogni palla con i rettangoli contenuti in essa che ricoprono lo stesso punto), e viceversa: le due topologie coincidono. **Il punto metodologico:** basi diverse possono generare **la stessa** topologia — la topologia è l'oggetto essenziale, la base è solo uno strumento descrittivo, e non è unica.

```python
import sympy as sp

# Verifica simbolica: il quadrato inscritto in una palla ha diagonale = r (raggio della palla)
r, lato = sp.symbols('r lato', positive=True)
# lato del quadrato = r/sqrt(2) => diagonale = lato*sqrt(2) = r
diagonale = (r / sp.sqrt(2)) * sp.sqrt(2)
print("Diagonale del quadrato di lato r/sqrt(2):", sp.simplify(diagonale), " (deve essere r)")

# Verifica numerica che il quadrato stia dentro la palla
import numpy as np
x0 = np.array([0.0, 0.0])
raggio = 2.0
lato_num = raggio / np.sqrt(2) * 0.999  # leggermente piu' piccolo per stare nella palla APERTA
angoli_test = [np.array([lato_num, lato_num]), np.array([-lato_num, lato_num]),
               np.array([lato_num, -lato_num]), np.array([-lato_num, -lato_num])]
for angolo in angoli_test:
    distanza = np.linalg.norm(x0 + angolo - x0)
    print(f"Vertice a distanza {distanza:.4f} dal centro (raggio palla={raggio}): dentro? {distanza < raggio}")
```

## Riepilogo finale

- Una **base** $\mathcal B$ di una topologia $\tau$ è una sottofamiglia di aperti che genera tutti gli altri per unione: descrive $\tau$ in modo compatto, senza elencarne ogni elemento.
- Una famiglia $\mathcal B$ genera una topologia se e solo se copre $X$ e le intersezioni dei suoi elementi contengono sempre un elemento più piccolo della base attorno a ogni punto.
- Basi diverse (palle, rettangoli) possono generare **la stessa** topologia: la topologia, non la base, è l'oggetto essenziale.
- Uno spazio a **base numerabile** (come $\mathbb R^n$, con le palle a centro razionale e raggio razionale) permette di ridurre argomenti su infiniti aperti arbitrari ad argomenti su una famiglia numerabile, spesso molto più maneggevole.
