## Perché ci interessa

Nel notebook 12.12 abbiamo enunciato, senza dimostrarlo, che $\pi_1(S^1)\cong\mathbb Z$: l'idea intuitiva era il "numero di avvolgimento" di un cappio, ma non avevamo gli strumenti per definirlo con precisione. Il concetto mancante è quello di **rivestimento**: uno spazio che "si arrotola" su un altro, localmente indistinguibile da esso ma globalmente diverso — l'esempio archetipico è la retta $\mathbb R$ che si arrotola su $S^1$ tramite $t\mapsto(\cos2\pi t,\sin2\pi t)$. Questo notebook, capstone del modulo di topologia, introduce i rivestimenti, dimostra il teorema di **sollevamento dei cammini** (lo strumento tecnico decisivo) e lo usa per calcolare finalmente, con una dimostrazione completa, $\pi_1(S^1)\cong\mathbb Z$.

## Teoria, passo dopo passo

### Definizione di rivestimento

Una funzione continua e suriettiva $p:\tilde X\to X$ è un **rivestimento** se ogni $x\in X$ ha un intorno aperto $U$ **ben ricoperto**: $p^{-1}(U)$ è unione disgiunta di aperti $\{V_i\}_{i\in I}$ di $\tilde X$, ciascuno mandato da $p$ **omeomorficamente** su $U$ (notebook 12.11). Intuitivamente: localmente, $\tilde X$ sopra $U$ appare come "tante copie parallele" di $U$, senza intrecci.

**Esempio fondamentale.** $p:\mathbb R\to S^1$, $p(t)=(\cos2\pi t,\sin2\pi t)$, è un rivestimento. Dato un punto $x_0=(\cos2\pi t_0,\sin2\pi t_0)\in S^1$, l'intorno $U$ = $S^1$ meno il punto antipodale a $x_0$ (un arco aperto che copre "quasi tutta" la circonferenza) ha controimmagine $p^{-1}(U)=\bigcup_{n\in\mathbb Z}(t_0-\tfrac12+n,\,t_0+\tfrac12+n)$: un'unione disgiunta di intervalli aperti (uno per ogni intero $n$, ciascuno di lunghezza $1$), e $p$ ristretta a ciascuno di essi è un omeomorfismo su $U$ (biettiva su un periodo, continua con inversa continua, per lo stesso argomento — ma ora corretto, perché stiamo usando un intervallo *aperto* invece del semiaperto $[0,2\pi)$ — del notebook 12.11). Questo rivestimento è, precisamente, il quoziente $\mathbb R\to\mathbb R/\!\sim$ del notebook 12.10 (con $\sim$ dato dall'esercizio 1 di quel notebook), riletto come mappa di rivestimento.

### Sollevamento di cammini

**Teorema (sollevamento dei cammini).** Sia $p:\tilde X\to X$ un rivestimento, $\gamma:[0,1]\to X$ un cammino con $\gamma(0)=x_0$, e $\tilde x_0\in p^{-1}(x_0)$ un punto scelto sopra $x_0$. Allora esiste un **unico** cammino $\tilde\gamma:[0,1]\to\tilde X$ (il **sollevamento**) con $p\circ\tilde\gamma=\gamma$ e $\tilde\gamma(0)=\tilde x_0$.

**Dimostrazione (schema, per compattezza omettendo alcuni dettagli tecnici sulla scelta della partizione).** Per ogni $s\in[0,1]$, $\gamma(s)$ ha un intorno ben ricoperto $U_s$; per continuità di $\gamma$, esiste un intervallo $(s-\delta_s,s+\delta_s)$ con $\gamma((s-\delta_s,s+\delta_s))\subseteq U_s$. La famiglia $\{(s-\delta_s,s+\delta_s)\}_{s\in[0,1]}$ ricopre $[0,1]$; per compattezza (Heine-Borel, notebook 10.11/12.6), esiste un sottoricoprimento finito, da cui si estrae una partizione $0=s_0<s_1<\cdots<s_n=1$ tale che $\gamma([s_{k-1},s_k])$ sta interamente in un aperto ben ricoperto $U_k$, per ogni $k$. Costruiamo $\tilde\gamma$ induttivamente: su $[s_0,s_1]=[0,s_1]$, essendo $p^{-1}(U_1)$ un'unione disgiunta di copie omeomorfe di $U_1$, esiste esattamente una copia $V_1$ contenente $\tilde x_0$; definiamo $\tilde\gamma=(p|_{V_1})^{-1}\circ\gamma$ su $[0,s_1]$ (ben definita e continua, essendo $p|_{V_1}$ un omeomorfismo). Procediamo allo stesso modo su $[s_1,s_2]$, partendo dal punto $\tilde\gamma(s_1)$ già determinato, e così via fino a $[s_{n-1},1]$. L'unicità segue dal fatto che, a ogni passo, la scelta della copia $V_k$ è **forzata** dal punto di partenza $\tilde\gamma(s_{k-1})$ già fissato (essendo le copie disgiunte). $\blacksquare$

**Il ruolo della compattezza:** questa dimostrazione usa direttamente il teorema di Heine-Borel (notebook 10.11, 12.6) per garantire che serva solo un numero **finito** di passi induttivi — un altro esempio, dopo quelli del Modulo 10 e 12.6, di come la compattezza trasformi un controllo "punto per punto" in un controllo gestibile in un numero finito di passi.

### Calcolo di $\pi_1(S^1)$

Definiamo $\Phi:\pi_1(S^1,1)\to\mathbb Z$ (dove $1=(1,0)$ è il punto base) così: dato un cappio $\gamma$ basato in $1$, sia $\tilde\gamma$ il suo sollevamento (unico per il teorema) con $\tilde\gamma(0)=0\in\mathbb R$ (scegliendo $0\in p^{-1}(1)$); poniamo $\Phi([\gamma])=\tilde\gamma(1)$ — che è **necessariamente un intero**, perché $p(\tilde\gamma(1))=\gamma(1)=1$ (essendo $\gamma$ un cappio), e $p^{-1}(1)=\mathbb Z$ (i punti $t$ con $\cos2\pi t=1,\sin2\pi t=0$ sono esattamente gli interi).

**Teorema (enunciato, dimostrazione completa fuori portata di un notebook introduttivo, ma la struttura è delineata).** $\Phi$ è ben definita (non dipende dal rappresentante della classe di omotopia — si dimostra sollevando l'intera omotopia $H:[0,1]\times[0,1]\to S^1$ allo stesso modo dei cammini, usando di nuovo compattezza), è un **omomorfismo di gruppi** ($\Phi([\gamma]*[\delta])=\Phi([\gamma])+\Phi([\delta])$: sollevare la concatenazione equivale a concatenare i sollevamenti, shiftando il secondo dell'ammontare del primo), ed è **biettiva** (iniettiva: due cappi con lo stesso numero di avvolgimento hanno sollevamenti con lo stesso punto finale, quindi sono omotopi tramite un'omotopia rettilinea sui sollevamenti in $\mathbb R$, poi proiettata via $p$; suriettiva: per ogni $n\in\mathbb Z$, il cappio $\gamma_n(s)=(\cos2\pi ns,\sin2\pi ns)$ ha sollevamento $\tilde\gamma_n(s)=ns$, con $\Phi([\gamma_n])=n$). Per il notebook 11.4, un omomorfismo biettivo è un isomorfismo: $\pi_1(S^1,1)\cong\mathbb Z$. $\blacksquare$

Questo chiude il cerchio aperto nel notebook 12.12: il "numero di avvolgimento", ora definito con precisione come il valore finale del sollevamento in $\mathbb R$, è esattamente l'isomorfismo che identifica $\pi_1(S^1)$ con $(\mathbb Z,+)$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Visualizziamo il sollevamento: un cappio su S^1 con avvolgimento 2, e il suo sollevamento in R
s_vals = np.linspace(0, 1, 200)
n_avvolgimento = 2
gamma = np.stack([np.cos(2*np.pi*n_avvolgimento*s_vals), np.sin(2*np.pi*n_avvolgimento*s_vals)], axis=1)
gamma_sollevato = n_avvolgimento * s_vals  # sollevamento: t(s) = n*s

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

axes[0].plot(gamma[:,0], gamma[:,1], 'b-', linewidth=2)
axes[0].plot(1, 0, 'ro', markersize=10, label='punto base (1,0)')
axes[0].set_aspect('equal')
axes[0].set_title(f"Cappio su S^1 con avvolgimento n={n_avvolgimento}")
axes[0].legend()

axes[1].plot(s_vals, gamma_sollevato, 'g-', linewidth=2)
axes[1].axhline(0, color='gray', linestyle=':')
axes[1].axhline(n_avvolgimento, color='red', linestyle='--', label=f'sollevamento finale = {n_avvolgimento}')
axes[1].set_xlabel("s")
axes[1].set_ylabel("sollevamento in R")
axes[1].set_title("Sollevamento del cappio: parte da 0, arriva a n")
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"Phi([gamma]) = sollevamento(1) = {gamma_sollevato[-1]:.1f} = numero di avvolgimento")
```

## Esercizio svolto e commentato

**Problema:** verifica esplicitamente, con il teorema di sollevamento, che il cappio $\gamma(s)=(\cos4\pi s,\sin4\pi s)$ (avvolgimento atteso $2$) e il cappio $\delta(s)=(\cos4\pi s,-\sin4\pi s)$ (stesso numero di giri ma verso opposto, avvolgimento atteso $-2$) hanno $\Phi([\gamma])=2$ e $\Phi([\delta])=-2$, e verifica che $\Phi([\gamma]*[\delta])=\Phi([\gamma])+\Phi([\delta])=0$, coerentemente con l'essere $\gamma*\delta$ omotopo al cappio costante.

**Ragionamento:**

**Passo 1 — sollevamento di $\gamma$.** Cerchiamo $\tilde\gamma$ con $p(\tilde\gamma(s))=\gamma(s)$, cioè $\cos(2\pi\tilde\gamma(s))=\cos4\pi s$ e $\sin(2\pi\tilde\gamma(s))=\sin4\pi s$: la scelta continua con $\tilde\gamma(0)=0$ è $\tilde\gamma(s)=2s$ (si verifica per sostituzione diretta). $\Phi([\gamma])=\tilde\gamma(1)=2$.

**Passo 2 — sollevamento di $\delta$.** Analogamente, $\delta(s)=(\cos(-4\pi s),\sin(-4\pi s))$ (usando $-\sin=\sin(-\cdot)$, parità/disparità di seno e coseno), quindi il sollevamento con $\tilde\delta(0)=0$ è $\tilde\delta(s)=-2s$. $\Phi([\delta])=\tilde\delta(1)=-2$.

**Passo 3 — sollevamento della concatenazione.** $\gamma*\delta$ percorre $\gamma$ raddoppiando la velocità sulla prima metà, poi $\delta$ sulla seconda: il sollevamento parte da $0$, sale linearmente fino a $\tilde\gamma(1)=2$ a metà percorso (essendo $\gamma$ già un cappio completo, il sollevamento di $\gamma$ raggiunge $2$), poi da lì segue il sollevamento di $\delta$ **shiftato** di $+2$ (perché il sollevamento della concatenazione riparte da dove si trovava, non da $0$): scende da $2$ fino a $2+(-2)=0$. Dunque il sollevamento completo di $\gamma*\delta$ va da $0$ a $0$: $\Phi([\gamma]*[\delta])=0=\Phi([\gamma])+\Phi([\delta])=2+(-2)$. ✓

**Conclusione:** $\Phi([\gamma*\delta])=0=\Phi([e])$ (il cappio costante ha sollevamento costante, quindi $\Phi([e])=0$); per l'iniettività di $\Phi$ (enunciata nella teoria), questo implica $[\gamma*\delta]=[e]$, cioè $\gamma*\delta$ è omotopo al cappio costante — un fatto tutt'altro che ovvio "a occhio" (il cammino $\gamma*\delta$ è tutt'altro che banale, facendo due giri avanti e due indietro), ma che l'omomorfismo $\Phi$ rende immediato da verificare tramite semplice aritmetica su $\mathbb Z$.

```python
import numpy as np

def solleva(coeff, s_vals):
    # solleva il cappio (cos(2*pi*coeff*s), sin(2*pi*coeff*s)) partendo da 0
    return coeff * s_vals

s = np.linspace(0, 1, 5)
sollevamento_gamma = solleva(2, s)
sollevamento_delta = solleva(-2, s)

print("Sollevamento di gamma (avvolgimento 2):", sollevamento_gamma)
print("Sollevamento di delta (avvolgimento -2):", sollevamento_delta)
print(f"\nPhi([gamma]) = {sollevamento_gamma[-1]}")
print(f"Phi([delta]) = {sollevamento_delta[-1]}")
print(f"Somma attesa per la concatenazione: {sollevamento_gamma[-1] + sollevamento_delta[-1]}")
print("\n-> gamma*delta ha avvolgimento netto 0: e' omotopo al cappio costante")
```

## Riepilogo finale

- Un **rivestimento** $p:\tilde X\to X$ è localmente un omeomorfismo (ogni punto ha un intorno le cui controimmagini sono copie disgiunte omeomorfe all'intorno), pur non essendo globalmente iniettivo — l'esempio chiave è $\mathbb R\to S^1$.
- Il **teorema di sollevamento dei cammini** (dimostrato usando la compattezza di $[0,1]$, esattamente come Heine-Borel nel Modulo 10) garantisce che ogni cammino sulla base si "solleva" in modo unico, fissato il punto di partenza sopra.
- Questo teorema permette di definire con precisione il numero di avvolgimento e dimostrare rigorosamente $\pi_1(S^1)\cong\mathbb Z$, l'isomorfismo lasciato in sospeso nel notebook 12.12.
