# 12.10 — Spazi quoziente

**Prerequisiti:** 12.9 Spazi prodotto, 10.3 Insiemi, relazioni, funzioni, 11.3 Gruppi quoziente.


## Perché ci interessa

Nel notebook 11.3 abbiamo costruito i gruppi quoziente "incollando" elementi di un gruppo tramite una relazione di equivalenza (i laterali di un sottogruppo normale), ottenendo una nuova struttura algebrica. La stessa idea — prendere un insieme di partenza e "incollare" punti equivalenti — funziona anche in topologia, e produce alcuni degli spazi più interessanti e controintuitivi: una circonferenza si ottiene incollando gli estremi di un segmento, un nastro di Möbius incollando i lati di un rettangolo con un "giro", un toro incollando entrambe le coppie di lati opposti. Lo **spazio quoziente** formalizza precisamente questa operazione di "incollamento", definendo su di essa una topologia coerente.

## Teoria, passo dopo passo

### Relazioni di equivalenza e proiezione al quoziente

Data una relazione di equivalenza $\sim$ su $X$ (riflessiva, simmetrica, transitiva — notebook 10.3), l'insieme quoziente $X/\!\sim$ è l'insieme delle classi di equivalenza $[x]=\{y\in X:y\sim x\}$. La **proiezione canonica** $q:X\to X/\!\sim$, $q(x)=[x]$, manda ogni punto nella propria classe — esattamente come nel notebook 11.3, dove $q:G\to G/N$ mandava ogni elemento nel proprio laterale.

### Topologia quoziente

Data una topologia $\tau$ su $X$, la **topologia quoziente** su $X/\!\sim$ è definita da: $U\subseteq X/\!\sim$ è aperto se e solo se $q^{-1}(U)$ è aperto in $X$. Si verifica che questa è effettivamente una topologia (gli assiomi si trasferiscono direttamente, essendo la controimmagine compatibile con unioni e intersezioni, notebook 10.3), ed è **la più fine** (con più aperti) che rende $q$ continua.

**Perché "più fine possibile": la definizione non è arbitraria.** Se volessimo che $q$ fosse continua (requisito minimo perché il quoziente "erediti" informazione topologica da $X$), servirebbe che $q^{-1}(U)$ fosse aperto per ogni aperto $U$ di $X/\!\sim$ (notebook 12.4): la topologia quoziente prende **come aperti tutti e soli** gli insiemi per cui questo è automaticamente vero, cioè il massimo possibile compatibilmente con la continuità di $q$.

### Esempio: la circonferenza come quoziente di un intervallo

Su $X=[0,1]$, definiamo $\sim$ identificando solo gli estremi: $0\sim1$, e ogni altro punto equivalente solo a se stesso. Lo spazio quoziente $[0,1]/\!\sim$ è **omeomorfo** alla circonferenza $S^1$ (concetto di omeomorfismo che formalizzeremo nel notebook 12.11): intuitivamente, si "piega" il segmento $[0,1]$ e si incollano i due estremi. Un aperto attorno alla classe $[0]=[1]$ nel quoziente corrisponde, tramite $q^{-1}$, a un aperto in $[0,1]$ che contiene **sia** un intorno di $0$ **sia** un intorno di $1$ — riflettendo il fatto che sulla circonferenza, il punto "cucitura" ha vicini da entrambi i lati del taglio originale.

### Esempio: nastro di Möbius come quoziente di un rettangolo

Su $X=[0,1]\times[0,1]$, la relazione $\sim$ identifica $(0,y)\sim(1,1-y)$ per ogni $y$ (incollando il lato sinistro al lato destro, ma con un **giro**: il basso si incolla all'alto e viceversa), lasciando gli altri punti equivalenti solo a se stessi. Il quoziente risultante è il **nastro di Möbius**: una superficie con una sola faccia e un solo bordo, dimostrabile percorrendo il bordo del rettangolo e osservando come l'identificazione lo trasforma in un'unica curva chiusa (invece delle due curve separate che si avrebbero incollando senza giro, $(0,y)\sim(1,y)$, ottenendo un cilindro).

### Continuità delle funzioni dal quoziente

**Teorema (proprietà universale del quoziente).** Una funzione $f:X/\!\sim\,\to Z$ è continua se e solo se $f\circ q:X\to Z$ è continua.

**Dimostrazione.** *(⇒)* Se $f$ è continua, $f\circ q$ è composizione di continue (notebook 12.4: $q$ è continua per costruzione della topologia quoziente). *(⇐)* Supponiamo $f\circ q$ continua. Sia $W$ aperto in $Z$: $(f\circ q)^{-1}(W)=q^{-1}(f^{-1}(W))$ è aperto in $X$ (per continuità di $f\circ q$). Per **definizione** di topologia quoziente, $q^{-1}(f^{-1}(W))$ aperto significa esattamente che $f^{-1}(W)$ è aperto in $X/\!\sim$. Dunque $f^{-1}(W)$ è aperto per ogni aperto $W$: $f$ è continua. $\blacksquare$ Questo teorema è lo strumento pratico principale: per definire una funzione continua sul quoziente, basta definirla su $X$ in modo **compatibile con $\sim$** (cioè costante su ogni classe — esattamente la "buona posizione" verificata nel notebook 11.1 per l'operazione su $\mathbb Z/n\mathbb Z$) e verificarne la continuità su $X$, molto più semplice da controllare direttamente.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

# Segmento [0,1] -> circonferenza, incollando 0 e 1
t = np.linspace(0, 1, 100)
axes[0].plot(t, np.zeros_like(t), 'b-', linewidth=3)
axes[0].plot([0, 1], [0, 0], 'ro', markersize=10)
axes[0].annotate("0 ~ 1\n(stesso punto\nnel quoziente)", xy=(0.5, 0.05), ha='center')
axes[0].set_title("[0,1] con 0~1 -> S^1 (circonferenza)")
axes[0].set_ylim(-0.3, 0.3)
axes[0].set_yticks([])

# La circonferenza risultante
theta = np.linspace(0, 2*np.pi, 100)
axes[1].plot(np.cos(theta), np.sin(theta), 'g-', linewidth=3)
axes[1].plot(1, 0, 'ro', markersize=10)
axes[1].annotate("[0]=[1]", xy=(1, 0), xytext=(1.3, 0.3))
axes[1].set_aspect('equal')
axes[1].set_title("Risultato: circonferenza S^1")

plt.tight_layout()
plt.show()

print("Un intorno di [0]=[1] nel quoziente contiene punti vicini a 0 E punti vicini a 1 in [0,1]")
print("q^-1(intorno di [0]) = (intervallo vicino a 0) unito (intervallo vicino a 1), aperto in [0,1]")
```

## Esercizio svolto e commentato

**Problema:** dimostra, usando la proprietà universale del quoziente, che $f:[0,1]/\!\sim\,\to S^1$ definita da $f([t])=(\cos2\pi t,\sin2\pi t)$ (dove $\sim$ identifica $0\sim1$ come nella teoria) è ben definita e continua.

**Ragionamento:**

**Passo 1 — verifichiamo che $f$ è ben definita.** $f$ deve dare lo stesso valore su punti equivalenti: $[0]=[1]$, e $f([0])=(\cos0,\sin0)=(1,0)$, mentre $f([1])$ sarebbe $(\cos2\pi,\sin2\pi)=(1,0)$: coincidono. Per ogni altro $t\in(0,1)$, $[t]=\{t\}$ (nessuna identificazione), quindi non c'è ambiguità. $f$ è ben definita.

**Passo 2 — costruiamo $g=f\circ q:[0,1]\to S^1$, $g(t)=(\cos2\pi t,\sin2\pi t)$.** Questa è la funzione "prima del passaggio al quoziente": per costruzione, $g=f\circ q$.

**Passo 3 — dimostriamo che $g$ è continua.** $g=(g_1,g_2)$ con $g_1(t)=\cos2\pi t$, $g_2(t)=\sin2\pi t$, entrambe continue su $[0,1]$ (funzioni elementari, continue su tutto $\mathbb R$, notebook 10.17). Per il teorema sulla continuità verso un prodotto (notebook 12.9), $g$ è continua (come funzione a valori in $\mathbb R^2$, e quindi anche come funzione a valori nel sottospazio $S^1\subseteq\mathbb R^2$, dato che la continuità verso un sottospazio equivale alla continuità verso lo spazio ambiente quando l'immagine sta nel sottospazio).

**Conclusione:** essendo $g=f\circ q$ continua, per il teorema della proprietà universale del quoziente (teoria), $f$ è continua. **Osservazione:** questa strategia — costruire prima una funzione continua su $X$ compatibile con $\sim$, poi dedurre la continuità sul quoziente — è **sempre** più semplice del tentativo diretto di lavorare con le classi di equivalenza, esattamente come per i gruppi quoziente (notebook 11.3) era più semplice verificare la buona posizione dell'operazione su rappresentanti che ragionare direttamente sulle classi.

```python
import numpy as np
import matplotlib.pyplot as plt

t_vals = np.linspace(0, 1, 100)
x_vals = np.cos(2 * np.pi * t_vals)
y_vals = np.sin(2 * np.pi * t_vals)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x_vals, y_vals, 'b-', linewidth=2)
ax.plot(x_vals[0], y_vals[0], 'ro', markersize=10, label='g(0) = g(1) = (1,0)')
ax.set_aspect('equal')
ax.legend()
ax.set_title("g(t) = (cos(2*pi*t), sin(2*pi*t)): g(0)=g(1), compatibile con 0~1")
plt.tight_layout()
plt.show()

print(f"g(0) = ({x_vals[0]:.6f}, {y_vals[0]:.6f})")
print(f"g(1) = ({x_vals[-1]:.6f}, {y_vals[-1]:.6f})")
print("Coincidono: g e' compatibile con la relazione 0~1, quindi induce una funzione ben definita sul quoziente")
```

**1.** Ogni classe $[x]=\{x+n:n\in\mathbb Z\}$ contiene esattamente un rappresentante in $[0,1)$: lo spazio quoziente $\mathbb R/\!\sim$ è in corrispondenza biunivoca con $[0,1)$, e la relazione "incolla" $0$ con tutti i punti $1,2,3,\ldots$ e $-1,-2,\ldots$, che sono tutti equivalenti a $0$. Topologicamente, $\mathbb R/\!\sim$ è omeomorfo alla circonferenza $S^1$: è, in sostanza, la stessa costruzione dell'esempio della teoria (dove si identificava solo $0$ con $1$ su $[0,1]$), ma ripetuta periodicamente su tutta la retta — un modo alternativo, spesso usato in analisi di Fourier, di pensare alla circonferenza come "la retta reale arrotolata".

**2.** Dato $[x]\in X/\!\sim$ (una qualunque classe di equivalenza), per definizione di insieme quoziente $[x]$ è la classe di un qualche elemento $x\in X$ (le classi di equivalenza, per costruzione, sono indicizzate dagli elementi di $X$, anche se più elementi possono dare la stessa classe). Allora $q(x)=[x]$: ogni elemento del quoziente è immagine di (almeno) un elemento di $X$ tramite $q$. Dunque $q$ è suriettiva.

**3.** $q:X\to X/\!\sim$ è continua (per costruzione della topologia quoziente) e suriettiva (esercizio 2). Per l'esercizio 1 del notebook 12.5 (l'immagine continua e suriettiva di un connesso è connessa), essendo $X$ connesso, $X/\!\sim\,=q(X)$ è connesso.

**4.** La relazione identifica il lato sinistro con il destro **senza giro** ($(0,y)\sim(1,y)$, stesso $y$, non $1-y$ come nel Möbius) e il lato inferiore con il superiore, anch'esso senza giro ($(x,0)\sim(x,1)$). Incollando prima una coppia di lati (senza giro) si ottiene un cilindro; incollando poi le due estremità circolari del cilindro (di nuovo senza giro) si ottiene il toro — una "ciambella". Il nastro di Möbius, invece, applica **un** giro nell'unica identificazione che compie: il numero di giri (pari o dispari) nelle identificazioni di un rettangolo è precisamente ciò che distingue superfici topologicamente diverse come cilindro/Möbius (una identificazione, 0 o 1 giro) e toro/bottiglia di Klein (due identificazioni, con 0 o 1 giro nella seconda) — un'anticipazione informale della classificazione delle superfici, argomento di topologia più avanzata.

## Riepilogo finale

- Lo **spazio quoziente** $X/\!\sim$ "incolla" i punti equivalenti secondo una relazione $\sim$; la topologia quoziente rende la proiezione $q$ continua nel modo più ricco possibile.
- Esempi paradigmatici: la circonferenza (segmento con estremi incollati), il nastro di Möbius (rettangolo con un lato incollato con un giro), il toro (rettangolo con entrambe le coppie di lati incollate senza giro).
- Per definire una funzione continua sul quoziente, basta costruire una funzione continua e compatibile con $\sim$ sullo spazio di partenza: la proprietà universale del quoziente traduce questo controllo locale in continuità globale.
- Proprietà come connessione si trasmettono da $X$ al quoziente, essendo $q$ sempre continua e suriettiva.
