# 12.8 — Completezza (spazi metrici)

**Prerequisiti:** 12.7 Spazi metrici, 10.9 Completezza, 10.10 Successioni rigorose.


## Perché ci interessa

Nel notebook 10.10 abbiamo dimostrato che in $\mathbb R$ ogni successione di Cauchy converge (il **criterio di Cauchy**), usando in modo essenziale l'assioma di completezza (notebook 10.9, sup/inf) e il teorema di Bolzano-Weierstrass. Ma la definizione di successione di Cauchy — i termini si avvicinano tra loro sempre di più — usa solo la **distanza**, non l'ordine né il sup: ha senso in **qualunque** spazio metrico. La completezza, riformulata così, diventa una proprietà che uno spazio metrico può avere o no, indipendentemente dall'esistenza di un ordine. Questo notebook generalizza Cauchy e completezza, e chiarisce un punto spesso frainteso: $\mathbb Q$ non è "incompleto" perché gli mancano dei numeri arbitrari, ma precisamente perché alcune sue successioni di Cauchy non hanno limite **dentro $\mathbb Q$**.

## Teoria, passo dopo passo

### Successioni di Cauchy in uno spazio metrico

Una successione $(x_n)$ in uno spazio metrico $(X,d)$ è di **Cauchy** se per ogni $\varepsilon>0$ esiste $N$ tale che $n,m\ge N\Rightarrow d(x_n,x_m)<\varepsilon$ — identica, parola per parola, alla definizione del notebook 10.10, con $d(x_n,x_m)$ al posto di $|x_n-x_m|$.

**Teorema.** Ogni successione **convergente** è di Cauchy (in ogni spazio metrico).

**Dimostrazione.** Se $x_n\to L$, per ogni $\varepsilon>0$ esiste $N$ con $d(x_n,L)<\varepsilon/2$ per $n\ge N$. Allora per $n,m\ge N$: $d(x_n,x_m)\le d(x_n,L)+d(L,x_m)<\varepsilon/2+\varepsilon/2=\varepsilon$ (disuguaglianza triangolare, notebook 12.7). $\blacksquare$ Questa direzione vale **sempre**, in ogni spazio metrico: è il viceversa — Cauchy $\Rightarrow$ convergente — che può fallire.

### Spazio metrico completo

$(X,d)$ è **completo** se ogni successione di Cauchy in $X$ converge a un punto di $X$.

**$\mathbb R$ (con la metrica standard) è completo:** dimostrato nel notebook 10.10, via Bolzano-Weierstrass (che a sua volta usa la completezza-sup del notebook 10.9). **$\mathbb Q$ (con la metrica ereditata da $\mathbb R$) non è completo:** la successione delle approssimazioni decimali di $\sqrt2$ ($1, 1.4, 1.41, 1.414,\ldots$, tutti razionali) è di Cauchy in $\mathbb Q$ (converge in $\mathbb R$ a $\sqrt2$, quindi soddisfa la condizione di Cauchy, che è puramente sulle distanze tra i termini — indipendente da *dove* converge), ma **non converge a nessun razionale**, perché $\sqrt2\notin\mathbb Q$ (dimostrato per assurdo con l'aritmetica di parità, notebook 10.6) e il limite in $\mathbb R$ è unico.

### $\mathbb R^n$ è completo

**Teorema.** $(\mathbb R^n,d_2)$ (metrica euclidea) è completo.

**Dimostrazione.** Sia $(x_k)$ di Cauchy in $\mathbb R^n$, con $x_k=(x_k^{(1)},\ldots,x_k^{(n)})$. Per ogni componente $i$, $|x_k^{(i)}-x_m^{(i)}|\le d_2(x_k,x_m)$ (ogni componente contribuisce al quadrato della distanza euclidea, quindi è dominata dalla distanza stessa): dunque $(x_k^{(i)})_k$ è di Cauchy in $\mathbb R$ per ogni $i$ fissato. Per completezza di $\mathbb R$ (notebook 10.10), ogni componente converge: $x_k^{(i)}\to L^{(i)}$. Ponendo $L=(L^{(1)},\ldots,L^{(n)})$, si verifica che $x_k\to L$ in $\mathbb R^n$ (la distanza euclidea $d_2(x_k,L)=\sqrt{\sum_i(x_k^{(i)}-L^{(i)})^2}\to0$, essendo somma finita di termini che tendono a $0$). $\blacksquare$ Questa dimostrazione mostra un fatto metodologicamente importante: la completezza di $\mathbb R^n$ si **riduce** a quella di $\mathbb R$, componente per componente — non serve un nuovo assioma, solo il notebook 10.10 applicato $n$ volte.

### Completamento di uno spazio metrico

Ogni spazio metrico incompleto $(X,d)$ può essere "completato": esiste uno spazio metrico completo $(\hat X,\hat d)$ e un'immersione $X\hookrightarrow\hat X$ che preserva le distanze, tale che $X$ è denso in $\hat X$ (ogni punto di $\hat X$ è limite di una successione di punti di $X$). $\mathbb R$ è, in questo senso preciso, **il completamento di $\mathbb Q$**: si costruisce identificando ogni successione di Cauchy di razionali con il numero reale a cui "vorrebbe convergere" (questa è, storicamente, una delle costruzioni rigorose di $\mathbb R$ a partire da $\mathbb Q$, alternativa a quella per sezioni di Dedekind spesso usata nel notebook 10.9).

```python
from fractions import Fraction

# Costruiamo l'approssimazione razionale di sqrt(2) via il metodo di Newton, restando sempre in Q
def newton_sqrt2(iterazioni):
    x = Fraction(1)  # punto di partenza razionale
    approssimazioni = [x]
    for _ in range(iterazioni):
        x = (x + Fraction(2) / x) / 2  # iterazione di Newton per sqrt(2), tutta in aritmetica razionale
        approssimazioni.append(x)
    return approssimazioni

approssimazioni = newton_sqrt2(6)
for i, a in enumerate(approssimazioni):
    print(f"x_{i} = {a} = {float(a):.10f}")

print(f"\nsqrt(2) reale = {2**0.5:.10f}")
print("La successione e' di Cauchy in Q (i termini si avvicinano tra loro),")
print("ma non converge a NESSUN razionale: il suo 'limite' sqrt(2) non appartiene a Q.")

# Verifica che la successione e' di Cauchy: le distanze tra termini successivi tendono a 0
for i in range(len(approssimazioni)-1):
    dist = abs(float(approssimazioni[i+1]) - float(approssimazioni[i]))
    print(f"d(x_{i+1}, x_{i}) = {dist:.2e}")
```

## Esercizio svolto e commentato

**Problema:** dimostra che $(0,1)$ con la metrica standard ereditata da $\mathbb R$ **non** è completo, esibendo esplicitamente una successione di Cauchy senza limite in $(0,1)$.

**Ragionamento:**

**Passo 1 — costruiamo la successione.** Consideriamo $x_n=1/n$ per $n\ge2$ (così $x_n\in(0,1)$ per ogni $n\ge2$, evitando $x_1=1\notin(0,1)$).

**Passo 2 — verifichiamo che è di Cauchy.** Per $n,m\ge N$: $|x_n-x_m|=|1/n-1/m|\le1/n+1/m\le2/N$. Dato $\varepsilon>0$, scegliendo $N>2/\varepsilon$, si ha $2/N<\varepsilon$, quindi $|x_n-x_m|<\varepsilon$ per ogni $n,m\ge N$: la successione è di Cauchy (identica alla verifica, con la stessa successione, che si potrebbe fare in $\mathbb R$).

**Passo 3 — verifichiamo che non converge in $(0,1)$.** In $\mathbb R$, $x_n=1/n\to0$ (limite noto, notebook 10.8). Per l'unicità del limite (già dimostrata nel Modulo 10 per successioni reali), se $(x_n)$ convergesse a un punto $L\in(0,1)$ **nella metrica di $(0,1)$** (che è la stessa metrica di $\mathbb R$, ristretta), convergerebbe anche in $\mathbb R$ allo stesso $L$ — ma il limite in $\mathbb R$ è $0$, e $0\notin(0,1)$. Dunque non esiste $L\in(0,1)$ a cui $(x_n)$ converge.

**Conclusione:** $(x_n)=(1/n)$ è di Cauchy in $(0,1)$ ma non converge a nessun punto di $(0,1)$: $(0,1)$ non è completo. **Osservazione:** questo è coerente con il fatto, visto nel notebook 12.6, che $(0,1)$ non è compatto — non è una coincidenza: vedremo (esercizio) che ogni sottoinsieme compatto di uno spazio metrico è sempre completo, quindi uno spazio metrico incompleto, come $(0,1)$, non può essere compatto.

```python
import numpy as np
import matplotlib.pyplot as plt

n_vals = np.arange(2, 50)
x_vals = 1 / n_vals

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(n_vals, x_vals, 'bo-', markersize=4)
ax.axhline(0, color='red', linestyle='--', label="limite 'candidato' = 0, FUORI da (0,1)")
ax.set_xlabel("n")
ax.set_ylabel("x_n = 1/n")
ax.set_title("Successione di Cauchy in (0,1) senza limite in (0,1)")
ax.legend()
plt.tight_layout()
plt.show()

# Verifica numerica della condizione di Cauchy
def verifica_cauchy(x_vals, epsilon):
    for N in range(len(x_vals)):
        if all(abs(x_vals[n] - x_vals[m]) < epsilon for n in range(N, len(x_vals)) for m in range(N, len(x_vals))):
            return N
    return None

for eps in [0.1, 0.01, 0.001]:
    N = verifica_cauchy(x_vals, eps)
    print(f"epsilon={eps}: N={N} funziona (per n,m>=N, |x_n-x_m|<epsilon)" if N else f"epsilon={eps}: N non trovato nel range")
```

Poniamo $M=\max(d(x_1,x_1),d(x_2,x_1),\ldots,d(x_{N-1},x_1),d(x_N,x_1)+1)$ (massimo di un numero **finito** di valori, quindi esiste). Per $n<N$: $d(x_n,x_1)\le M$ per costruzione. Per $n\ge N$: $d(x_n,x_1)\le d(x_n,x_N)+d(x_N,x_1)<1+d(x_N,x_1)\le M$ (disuguaglianza triangolare). Dunque $d(x_n,x_1)\le M$ per ogni $n$: la successione è limitata.

**4.** Sia $(x_n)$ di Cauchy in $X$ compatto. Per compattezza (equivalente, in spazi metrici, alla compattezza per successioni — generalizzazione di Bolzano-Weierstrass), esiste una sottosuccessione $(x_{n_k})$ convergente a un punto $L\in X$. Mostriamo che l'**intera** successione converge a $L$: dato $\varepsilon>0$, per Cauchy esiste $N_1$ con $d(x_n,x_m)<\varepsilon/2$ per $n,m\ge N_1$; per la convergenza della sottosuccessione, esiste $k$ con $n_k\ge N_1$ e $d(x_{n_k},L)<\varepsilon/2$. Allora per ogni $n\ge N_1$: $d(x_n,L)\le d(x_n,x_{n_k})+d(x_{n_k},L)<\varepsilon/2+\varepsilon/2=\varepsilon$ (disuguaglianza triangolare, usando $n,n_k\ge N_1$ per il primo termine). Dunque $x_n\to L\in X$: $(X,d)$ è completo.

## Riepilogo finale

- Una successione di Cauchy in uno spazio metrico è definita esattamente come nel notebook 10.10, con la distanza $d$ al posto del valore assoluto; ogni successione convergente è di Cauchy, ma non sempre vale il viceversa.
- Uno spazio metrico è **completo** se ogni successione di Cauchy converge in esso: $\mathbb R$ e $\mathbb R^n$ sono completi, $\mathbb Q$ e $(0,1)$ non lo sono.
- La completezza di $\mathbb R^n$ si riduce, componente per componente, a quella di $\mathbb R$.
- Ogni spazio compatto è completo; ogni sottospazio chiuso di uno spazio completo è completo — collegando completezza, chiusura e compattezza in un quadro coerente.
