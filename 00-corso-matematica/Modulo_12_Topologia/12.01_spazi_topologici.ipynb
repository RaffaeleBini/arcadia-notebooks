## Perché ci interessa

Nel notebook 10.12 abbiamo definito la continuità in $\mathbb R$ con $\varepsilon$ e $\delta$: $f$ è continua in $x_0$ se per ogni $\varepsilon>0$ esiste $\delta>0$ tale che $|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$. Questa definizione usa il **valore assoluto**, cioè una nozione di distanza. Ma la continuità, intuitivamente, è un'idea più generale: "punti vicini vanno in punti vicini", e questa idea ha senso anche dove non c'è una distanza numerica — per esempio per le funzioni tra spazi di funzioni (notebook 10.16, convergenza uniforme), o per capire quando due figure geometriche sono "essenzialmente la stessa forma" senza misurare lunghezze. La **topologia** isola l'essenza della "vicinanza" — gli **insiemi aperti** — dimenticando la distanza stessa. Il guadagno, come per i gruppi nel notebook 11.1, è che teoremi dimostrati una volta per gli spazi topologici valgono automaticamente ovunque questa struttura minimale sia presente.

## Teoria, passo dopo passo

### Cosa rende speciali gli intervalli aperti di $\mathbb R$?

In $\mathbb R$, un insieme $U$ è **aperto** se ogni suo punto ha attorno un intervallo $(x-\delta,x+\delta)\subseteq U$ interamente contenuto in $U$ (nessun punto è "sul bordo"). Osserviamo tre fatti, verificabili direttamente dalla definizione:

1. $\emptyset$ e $\mathbb R$ sono aperti (banalmente: la condizione su $\emptyset$ è vuotamente vera; per $\mathbb R$ ogni $\delta$ va bene).
2. L'**unione** di una famiglia qualsiasi (anche infinita) di aperti è aperta: se $x\in\bigcup_i U_i$, allora $x\in U_j$ per qualche $j$, e l'intervallo attorno a $x$ dentro $U_j$ sta anche dentro $\bigcup_i U_i$.
3. L'**intersezione di un numero finito** di aperti è aperta: se $x\in U_1\cap\cdots\cap U_n$, ogni $U_i$ fornisce un $\delta_i$; prendendo $\delta=\min(\delta_1,\ldots,\delta_n)>0$ (minimo di un numero **finito** di positivi, quindi ancora positivo — qui serve la finitezza!) si ottiene un intervallo dentro tutti gli $U_i$ contemporaneamente.

L'intersezione infinita **non** conserva l'apertura in generale: $\bigcap_{n=1}^\infty(-1/n,1/n)=\{0\}$, che non è aperto (nessun intervallo attorno a $0$ sta dentro $\{0\}$). Questi tre fatti — non la distanza in sé — sono ciò che rende $\mathbb R$ uno "spazio dove ha senso parlare di vicinanza". La topologia li prende come **assiomi**.

### Definizione di spazio topologico

Uno **spazio topologico** è una coppia $(X,\tau)$, dove $X$ è un insieme e $\tau\subseteq\mathcal P(X)$ (l'insieme delle parti di $X$, cioè l'insieme di tutti i suoi sottoinsiemi) è una famiglia di sottoinsiemi di $X$, detti **aperti**, tale che:

1. $\emptyset,X\in\tau$;
2. l'unione di una famiglia qualsiasi di elementi di $\tau$ appartiene a $\tau$;
3. l'intersezione di un numero **finito** di elementi di $\tau$ appartiene a $\tau$.

$\tau$ si chiama **topologia** su $X$. Nota l'asimmetria tra (2) e (3): è voluta, e riflette esattamente il controesempio $\bigcap_n(-1/n,1/n)=\{0\}$ appena visto.

### Esempi fondamentali

- **Topologia euclidea (o standard) su $\mathbb R$**: $\tau$ = tutti gli insiemi che sono unione di intervalli aperti. È la topologia "naturale", quella implicita nel notebook 10.12.
- **Topologia discreta**: $\tau=\mathcal P(X)$, ogni sottoinsieme è aperto. È la topologia "più ricca possibile": ogni punto è isolato dagli altri.
- **Topologia indiscreta (banale)**: $\tau=\{\emptyset,X\}$. È la topologia "più povera possibile": nessuna distinzione tra i punti.
- **Topologia euclidea su $\mathbb R^n$**: generata dalle palle aperte $B(x,r)=\{y\in\mathbb R^n:\|y-x\|<r\}$, dove $\|\cdot\|$ è la norma euclidea del notebook 6.20 (radice quadrata della somma dei quadrati delle componenti) — la generalizzazione naturale del valore assoluto.

### Intorni

Un **intorno** di $x\in X$ è un insieme $V$ che contiene un aperto contenente $x$ (in molti testi, semplicemente un aperto contenente $x$: le due convenzioni sono equivalenti ai fini pratici). Gli intorni formalizzano "vicino a $x$" nel linguaggio della teoria: la definizione $\varepsilon$-$\delta$ del notebook 10.8 diceva, in sostanza, che per ogni intorno $(f(x_0)-\varepsilon,f(x_0)+\varepsilon)$ del valore $f(x_0)$ esiste un intorno di $x_0$ la cui immagine ci sta dentro — vedremo nel notebook 12.4 che questa è **esattamente** la definizione topologica di continuità, riscritta con gli intorni al posto di $\varepsilon,\delta$.

```python
import matplotlib.pyplot as plt
import numpy as np

# Visualizziamo perche' l'intersezione infinita di aperti puo' non essere aperta
fig, ax = plt.subplots(figsize=(7, 2))
for n in [1, 2, 3, 5, 10, 30]:
    ax.plot([-1/n, 1/n], [n, n], marker='|', markersize=10, color='steelblue')
ax.axvline(0, color='red', linestyle='--', label='x=0 (unico punto dell\'intersezione)')
ax.set_yticks([])
ax.set_xlabel("x")
ax.set_title("Intervalli (-1/n, 1/n): la loro intersezione infinita e' {0}, non aperta")
ax.legend()
plt.tight_layout()
plt.show()

# Verifica diretta: {0} non contiene nessun intervallo (-delta, delta) con delta>0
# perche' per ogni delta>0, esiste sempre un punto x con 0<|x|<delta (es. x=delta/2)
delta = 0.001
x_test = delta / 2
print(f"Con delta={delta}, il punto x={x_test} sta in (-delta,delta) ma non in {{0}}: {x_test != 0}")
```

## Esercizio svolto e commentato

**Problema:** verifica che la famiglia $\tau=\{\emptyset,\{a\},\{a,b\},\{a,b,c\}\}$ è una topologia su $X=\{a,b,c\}$, ed elenca tutti gli aperti che contengono $b$.

**Ragionamento:**

**Passo 1 — verifica dell'assioma (1).** $\emptyset\in\tau$ per costruzione, e $X=\{a,b,c\}\in\tau$ per costruzione. ✓

**Passo 2 — verifica dell'assioma (2), unioni.** Con solo 4 elementi in $\tau$, controlliamo tutte le unioni possibili di sottofamiglie: $\{a\}\cup\{a,b\}=\{a,b\}\in\tau$; $\{a\}\cup\{a,b,c\}=\{a,b,c\}\in\tau$; $\{a,b\}\cup\{a,b,c\}=\{a,b,c\}\in\tau$; ogni unione con $\emptyset$ non cambia nulla. Tutte le unioni restano in $\tau$. ✓ (Il punto metodologico: essendo $\tau$ **totalmente ordinato per inclusione** in questo esempio — $\emptyset\subset\{a\}\subset\{a,b\}\subset X$ — ogni unione è semplicemente il più grande dei due insiemi, quindi automaticamente in $\tau$.)

**Passo 3 — verifica dell'assioma (3), intersezioni finite.** Per lo stesso motivo (catena totalmente ordinata), ogni intersezione è il più piccolo dei due insiemi, quindi ancora in $\tau$: per esempio $\{a,b\}\cap\{a,b,c\}=\{a,b\}\in\tau$. ✓

**Passo 4 — aperti contenenti $b$.** Scorrendo $\tau=\{\emptyset,\{a\},\{a,b\},\{a,b,c\}\}$: $b\notin\emptyset$, $b\notin\{a\}$, $b\in\{a,b\}$, $b\in\{a,b,c\}$. Gli aperti contenenti $b$ sono $\{a,b\}$ e $\{a,b,c\}$.

**Osservazione:** questo esempio mostra che una topologia può essere "asimmetrica" tra i punti: $a$ appartiene a *tre* aperti diversi ($\{a\},\{a,b\},\{a,b,c\}$), mentre $c$ appartiene a un solo aperto ($X$ stesso). In uno spazio come questo (chiamato **topologia delle catene** o simile), i punti non sono "topologicamente equivalenti" — a differenza della topologia discreta, dove ogni punto è isolato allo stesso modo.

```python
# Verifica computazionale degli assiomi di topologia su un insieme finito
from itertools import combinations

X = frozenset({'a', 'b', 'c'})
tau = [frozenset(), frozenset({'a'}), frozenset({'a', 'b'}), frozenset({'a', 'b', 'c'})]

def verifica_topologia(X, tau):
    ok = True
    if frozenset() not in tau or X not in tau:
        print("Fallisce l'assioma (1): mancano vuoto o X")
        ok = False
    # unioni di tutte le coppie
    for A, B in combinations(tau, 2):
        if (A | B) not in tau:
            print(f"Fallisce l'assioma (2): {A} unione {B} = {A|B} non e' in tau")
            ok = False
    # intersezioni di tutte le coppie
    for A, B in combinations(tau, 2):
        if (A & B) not in tau:
            print(f"Fallisce l'assioma (3): {A} intersezione {B} = {A&B} non e' in tau")
            ok = False
    return ok

print("La famiglia data e' una topologia su X?", verifica_topologia(X, tau))

# Aperti contenenti 'b'
aperti_con_b = [set(A) for A in tau if 'b' in A]
print("Aperti che contengono 'b':", aperti_con_b)
```

Per De Morgan (notebook 0.13), $X\setminus\bigcup_i U_i=\bigcap_i(X\setminus U_i)$. Ogni $X\setminus U_i$ è finito (per $U_i\ne\emptyset$); un'intersezione di insiemi, di cui almeno uno è finito, è finita (un sottoinsieme di un insieme finito è finito). Per induzione (notebook 10.4) questo si estende a un numero finito qualsiasi di insiemi. Dunque $\tau_{cof}$ è una topologia.

## Riepilogo finale

- Uno **spazio topologico** $(X,\tau)$ generalizza la nozione di "insieme aperto" di $\mathbb R$ (notebook 10.12) a un insieme qualsiasi, tramite tre assiomi: $\emptyset,X$ aperti, unioni qualsiasi di aperti aperte, intersezioni **finite** di aperti aperte.
- L'asimmetria tra unioni (qualsiasi) e intersezioni (finite) non è arbitraria: riflette un fatto vero già in $\mathbb R$, dove intersezioni infinite di aperti possono ridursi a un punto, non aperto.
- Esempi estremi: la topologia **discreta** (tutto è aperto) e quella **indiscreta** (solo $\emptyset$ e $X$ sono aperti) mostrano quanto sia flessibile la definizione.
- Un **intorno** di un punto formalizza "vicino a quel punto", ed è il linguaggio con cui riscriveremo la continuità (notebook 12.4) senza più bisogno di $\varepsilon$ e $\delta$.
