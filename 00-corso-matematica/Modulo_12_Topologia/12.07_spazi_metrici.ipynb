# 12.7 — Spazi metrici

**Prerequisiti:** 12.6 Compattezza (topologica), 10.8 Limite $\varepsilon$-$\delta$, 6.20 Prodotti scalari.


## Perché ci interessa

Nei notebook precedenti (12.1-12.6) abbiamo costruito la topologia in astratto, con aperti che soddisfano assiomi, senza mai usare numeri o distanze. Ma quasi tutti gli esempi concreti — $\mathbb R$, $\mathbb R^n$, gli spazi di funzioni del notebook 10.16 — nascono da una **distanza**: la topologia euclidea era definita tramite le palle $B(x,r)$, che presuppongono di saper misurare $\|y-x\|$. Uno **spazio metrico** rende esplicita questa struttura intermedia: più ricca di uno spazio topologico generico (perché c'è una distanza numerica), ma più semplice di uno spazio vettoriale normato (perché non serve una struttura lineare sottostante). Formalizzare gli assiomi della distanza permette di riconoscere che moltissimi oggetti diversi — numeri, punti del piano, funzioni, persino stringhe di testo — vivono naturalmente in uno spazio metrico, ereditando automaticamente tutta la teoria topologica già sviluppata.

## Teoria, passo dopo passo

### Cosa rende "buona" una distanza?

Il valore assoluto $|x-y|$ su $\mathbb R$ (usato fin dal notebook 10.8 per gli $\varepsilon$-$\delta$) e la norma euclidea $\|x-y\|$ su $\mathbb R^n$ (notebook 6.20, derivata dal prodotto scalare come $\|v\|=\sqrt{v\cdot v}$) condividono proprietà precise: sono sempre $\ge0$, si annullano solo quando i due punti coincidono, sono simmetriche, e soddisfano la **disuguaglianza triangolare** ($|x-z|\le|x-y|+|y-z|$, dimostrata per la norma euclidea nel notebook 6.20 via Cauchy-Schwarz). Queste quattro proprietà, prese come assiomi, definiscono la nozione generale.

### Definizione di spazio metrico

Uno **spazio metrico** è una coppia $(X,d)$, con $X$ insieme e $d:X\times X\to\mathbb R$ (la **metrica** o **distanza**) tale che, per ogni $x,y,z\in X$:

1. $d(x,y)\ge0$, e $d(x,y)=0\iff x=y$ (positività, non-degenerazione);
2. $d(x,y)=d(y,x)$ (simmetria);
3. $d(x,z)\le d(x,y)+d(y,z)$ (disuguaglianza triangolare).

### Esempi

- $(\mathbb R,d)$ con $d(x,y)=|x-y|$: la metrica **standard**, quella implicita in tutto il notebook 10.8.
- $(\mathbb R^n,d)$ con $d(x,y)=\|x-y\|=\sqrt{\sum_i(x_i-y_i)^2}$: la metrica **euclidea**, che eredita la disuguaglianza triangolare dalla disuguaglianza di Cauchy-Schwarz (notebook 6.20).
- La **metrica discreta**: $d(x,y)=0$ se $x=y$, $d(x,y)=1$ altrimenti, su un $X$ qualsiasi. Verifica banale degli assiomi, ma genera la topologia **discreta** (ogni palla di raggio $<1$ è un singoletto, quindi ogni punto è aperto).
- La **metrica del taxicab** (o $\ell^1$) su $\mathbb R^2$: $d(x,y)=|x_1-y_1|+|x_2-y_2|$ (distanza percorsa muovendosi solo in orizzontale/verticale, come per le strade di una città a griglia) — diversa dalla euclidea come numero, ma vedremo (esercizio) che genera la **stessa** topologia.

### Ogni metrica induce una topologia

Definiamo la **palla aperta** $B(x,r)=\{y\in X:d(x,y)<r\}$. Le palle formano una base (verifica analoga a quella del notebook 12.2 per $\mathbb R^n$, usando la disuguaglianza triangolare al posto del calcolo diretto con i rettangoli): dato $y\in B(x_1,r_1)\cap B(x_2,r_2)$, ponendo $r=\min(r_1-d(x_1,y),r_2-d(x_2,y))>0$, si verifica (per la disuguaglianza triangolare) che $B(y,r)\subseteq B(x_1,r_1)\cap B(x_2,r_2)$. La topologia generata da questa base si chiama **topologia indotta dalla metrica**.

### Continuità metrica vs. topologica

**Teorema.** Siano $(X,d_X)$, $(Y,d_Y)$ spazi metrici, con le rispettive topologie indotte. $f:X\to Y$ è continua nel senso topologico (notebook 12.4) se e solo se, per ogni $x_0\in X$ e ogni $\varepsilon>0$, esiste $\delta>0$ tale che $d_X(x,x_0)<\delta\Rightarrow d_Y(f(x),f(x_0))<\varepsilon$ (continuità $\varepsilon$-$\delta$, generalizzazione diretta del notebook 10.8).

**Dimostrazione.** Segue immediatamente dall'equivalenza tra continuità globale e puntuale (notebook 12.4), osservando che gli intorni di base di $f(x_0)$ sono esattamente le palle $B(f(x_0),\varepsilon)$, e la condizione "$f(U)\subseteq B(f(x_0),\varepsilon)$ per qualche intorno $U$ di $x_0$" equivale a "esiste $\delta>0$ con $f(B(x_0,\delta))\subseteq B(f(x_0),\varepsilon)$", che è precisamente $d_X(x,x_0)<\delta\Rightarrow d_Y(f(x),f(x_0))<\varepsilon$. $\blacksquare$ Questo conferma che tutta la teoria $\varepsilon$-$\delta$ del Modulo 10 era, a posteriori, un caso particolare della continuità topologica, specializzata alla metrica standard di $\mathbb R$.

### Ogni spazio metrico è di Hausdorff

**Teorema.** Ogni spazio metrico $(X,d)$, con la topologia indotta, è di Hausdorff (definito nel notebook 12.6).

**Dimostrazione.** Dati $x\ne y$, sia $r=d(x,y)>0$ (positivo per l'assioma di non-degenerazione). Le palle $B(x,r/2)$ e $B(y,r/2)$ sono disgiunte: se $z$ stesse in entrambe, per la disuguaglianza triangolare $d(x,y)\le d(x,z)+d(z,y)<r/2+r/2=r$, assurdo essendo $d(x,y)=r$. $\blacksquare$ Questo giustifica, a posteriori, l'uso implicito di Hausdorff nel teorema "compatto $\Rightarrow$ chiuso" del notebook 12.6 applicato a $\mathbb R^n$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Confrontiamo visivamente le "palle" di tre metriche diverse su R^2: euclidea, taxicab, discreta (concettuale)
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

theta = np.linspace(0, 2*np.pi, 300)
r = 1

# Palla euclidea: cerchio
axes[0].plot(r*np.cos(theta), r*np.sin(theta), 'b-', linewidth=2)
axes[0].fill(r*np.cos(theta), r*np.sin(theta), alpha=0.3)
axes[0].set_title("Palla euclidea B(0,1): d(x,y) = ||x-y||")
axes[0].set_aspect('equal')

# Palla taxicab (norma L1): rombo |x|+|y| < 1
rombo_x = [1, 0, -1, 0, 1]
rombo_y = [0, 1, 0, -1, 0]
axes[1].plot(rombo_x, rombo_y, 'r-', linewidth=2)
axes[1].fill(rombo_x, rombo_y, alpha=0.3, color='red')
axes[1].set_title("Palla taxicab B(0,1): d(x,y) = |x1-y1|+|x2-y2|")
axes[1].set_aspect('equal')

for ax in axes:
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Verifica numerica della disuguaglianza triangolare per la metrica taxicab
def d_taxicab(p, q):
    return abs(p[0]-q[0]) + abs(p[1]-q[1])

x, y, z = np.array([0,0]), np.array([1,2]), np.array([3,1])
print(f"d(x,z) = {d_taxicab(x,z)}, d(x,y)+d(y,z) = {d_taxicab(x,y)+d_taxicab(y,z)}")
print("Disuguaglianza triangolare verificata:", d_taxicab(x,z) <= d_taxicab(x,y) + d_taxicab(y,z))
```

## Esercizio svolto e commentato

**Problema:** dimostra che la metrica euclidea $d_2(x,y)=\|x-y\|$ e la metrica taxicab $d_1(x,y)=|x_1-y_1|+|x_2-y_2|$ su $\mathbb R^2$ generano la **stessa** topologia.

**Ragionamento:**

**Passo 1 — strategia.** Per il criterio dell'esercizio 4 del notebook 12.2 (basi che generano topologie confrontabili), basta dimostrare che ogni palla euclidea contiene una palla taxicab attorno a ogni suo punto, e viceversa.

**Passo 2 — confronto quantitativo tra le due distanze.** Dimostriamo la doppia disuguaglianza $d_2(x,y)\le d_1(x,y)\le\sqrt2\,d_2(x,y)$. Per la prima: $d_2(x,y)^2=(x_1-y_1)^2+(x_2-y_2)^2\le(x_1-y_1)^2+(x_2-y_2)^2+2|x_1-y_1||x_2-y_2|=(|x_1-y_1|+|x_2-y_2|)^2=d_1(x,y)^2$ (il termine aggiunto è $\ge0$), quindi $d_2\le d_1$. Per la seconda: $(|x_1-y_1|+|x_2-y_2|)^2\le2(|x_1-y_1|^2+|x_2-y_2|^2)$ (disuguaglianza $(a+b)^2\le2(a^2+b^2)$, verificabile espandendo: equivale a $(a-b)^2\ge0$), quindi $d_1(x,y)\le\sqrt2\,d_2(x,y)$.

**Passo 3 — dalle disuguaglianze alle palle.** Dato $x_0$ e $r>0$, la disuguaglianza $d_2\le d_1$ dà: $d_1(x,x_0)<r\Rightarrow d_2(x,x_0)\le d_1(x,x_0)<r$, cioè $B_1(x_0,r)\subseteq B_2(x_0,r)$ (ogni palla taxicab di raggio $r$ è contenuta nella palla euclidea di raggio $r$). Dalla disuguaglianza $d_1\le\sqrt2\,d_2$: $d_2(x,x_0)<r/\sqrt2\Rightarrow d_1(x,x_0)\le\sqrt2\,d_2(x,x_0)<r$, cioè $B_2(x_0,r/\sqrt2)\subseteq B_1(x_0,r)$.

**Conclusione:** ogni palla euclidea $B_2(x_0,r)$ contiene la palla taxicab $B_1(x_0,r)$ (dal passo 3, con lo stesso raggio $r$), e ogni palla taxicab $B_1(x_0,r)$ contiene la palla euclidea $B_2(x_0,r/\sqrt2)$: le due basi si "raffinano a vicenda", quindi generano la stessa topologia. **Punto metodologico:** due metriche numericamente diverse ($d_1\ne d_2$ come funzioni) possono comunque produrre esattamente le stesse nozioni di aperto, continuità, convergenza — la topologia "dimentica" i dettagli quantitativi della metrica, trattenendo solo la struttura qualitativa di vicinanza.

```python
import numpy as np

def d2(x, y):
    return np.linalg.norm(np.array(x) - np.array(y))

def d1(x, y):
    return abs(x[0]-y[0]) + abs(x[1]-y[1])

# Verifica numerica della doppia disuguaglianza su punti casuali
np.random.seed(0)
for _ in range(5):
    x = np.random.randn(2)
    y = np.random.randn(2)
    v_d1, v_d2 = d1(x, y), d2(x, y)
    print(f"d2={v_d2:.3f}, d1={v_d1:.3f}, sqrt(2)*d2={np.sqrt(2)*v_d2:.3f} "
          f"-> d2<=d1<=sqrt(2)*d2 ? {v_d2 <= v_d1 + 1e-9 and v_d1 <= np.sqrt(2)*v_d2 + 1e-9}")
```

## Esercizi da risolvere

1. Verifica che la metrica discreta ($d(x,y)=1$ se $x\ne y$, $0$ altrimenti) soddisfa i tre assiomi di spazio metrico su un $X$ qualsiasi.
2. Dimostra che nella metrica discreta, la palla $B(x,1/2)=\{x\}$ (un singoletto): deduci che la topologia indotta dalla metrica discreta è la topologia discreta.
3. Dimostra che $d_\infty(x,y)=\max(|x_1-y_1|,|x_2-y_2|)$ (metrica del massimo, o **di Chebyshev**) su $\mathbb R^2$ è effettivamente una metrica (verifica i tre assiomi).
4. *(più impegnativo)* Dimostra che $d_\infty\le d_2\le\sqrt2\,d_\infty$ su $\mathbb R^2$, e deduci (con la stessa tecnica dell'esercizio svolto) che $d_\infty$ e $d_2$ generano la stessa topologia.

<summary><strong>👉 Soluzioni (clicca per visualizzare)</strong></summary>

**1.** *Positività*: $d(x,y)\ge0$ sempre (è $0$ o $1$); $d(x,y)=0\iff x=y$ per definizione. *Simmetria*: se $x\ne y$, anche $y\ne x$, quindi $d(x,y)=1=d(y,x)$; se $x=y$, $d(x,y)=0=d(y,x)$. *Disuguaglianza triangolare*: se $x=z$, $d(x,z)=0\le d(x,y)+d(y,z)$ banalmente (somma di non-negativi). Se $x\ne z$: $d(x,z)=1$; se fosse $d(x,y)=d(y,z)=0$, allora $x=y$ e $y=z$, quindi $x=z$, assurdo — dunque almeno uno tra $d(x,y),d(y,z)$ vale $1$, quindi $d(x,y)+d(y,z)\ge1=d(x,z)$.

**2.** $B(x,1/2)=\{y\in X:d(x,y)<1/2\}$: essendo $d$ a valori solo in $\{0,1\}$, la condizione $d(x,y)<1/2$ equivale a $d(x,y)=0$, cioè $y=x$. Dunque $B(x,1/2)=\{x\}$. Essendo ogni singoletto una palla aperta (quindi aperto), e ogni sottoinsieme unione di singoletti dei suoi punti, ogni sottoinsieme di $X$ è aperto nella topologia indotta: è la topologia discreta.

**3.** *Positività*: massimo di quantità $\ge0$ è $\ge0$; $d_\infty(x,y)=0\iff\max(|x_1-y_1|,|x_2-y_2|)=0\iff$ entrambi i termini sono $0$ (essendo non-negativi) $\iff x=y$. *Simmetria*: $|x_i-y_i|=|y_i-x_i|$ per ogni componente, quindi il massimo è lo stesso. *Disuguaglianza triangolare*: per ogni componente $i$, $|x_i-z_i|\le|x_i-y_i|+|y_i-z_i|\le d_\infty(x,y)+d_\infty(y,z)$ (ogni termine della somma è $\le$ del rispettivo massimo). Prendendo il massimo su $i=1,2$ a sinistra: $d_\infty(x,z)=\max_i|x_i-z_i|\le d_\infty(x,y)+d_\infty(y,z)$ (essendo il membro destro un maggiorante comune per entrambe le componenti).

**4.** *$d_\infty\le d_2$*: $d_\infty(x,y)^2=\max(|x_1-y_1|,|x_2-y_2|)^2\le|x_1-y_1|^2+|x_2-y_2|^2=d_2(x,y)^2$ (il massimo dei due quadrati è al più la loro somma, essendo entrambi $\ge0$), quindi $d_\infty\le d_2$. *$d_2\le\sqrt2\,d_\infty$*: $d_2(x,y)^2=|x_1-y_1|^2+|x_2-y_2|^2\le2\max(|x_1-y_1|,|x_2-y_2|)^2=2\,d_\infty(x,y)^2$ (ogni termine è al più il massimo dei due), quindi $d_2\le\sqrt2\,d_\infty$. Con lo stesso argomento dell'esercizio svolto (passo 3): $B_\infty(x_0,r)\subseteq B_2(x_0,r)$ e $B_2(x_0,r/\sqrt2)\subseteq B_\infty(x_0,r)$: le due basi si raffinano a vicenda, generando la stessa topologia.

## Riepilogo finale

- Uno **spazio metrico** $(X,d)$ ha una distanza $d$ che soddisfa positività, simmetria e disuguaglianza triangolare — la generalizzazione astratta di $|x-y|$ e $\|x-y\|$.
- Ogni metrica induce una topologia (le palle $B(x,r)$ formano una base), e la continuità $\varepsilon$-$\delta$ del Modulo 10 risulta essere un caso particolare della continuità topologica.
- Ogni spazio metrico è di Hausdorff: la distanza positiva tra punti distinti fornisce sempre palle disgiunte.
- Metriche numericamente diverse (euclidea, taxicab, del massimo) possono generare la **stessa** topologia, se limitate l'una dall'altra da costanti positive — la topologia trattiene solo l'informazione qualitativa, non quella quantitativa.
