# 12.5 — Connessione

**Prerequisiti:** 12.4 Continuità (topologica), 10.9 Completezza, 10.12 Continuità rigorosa.


## Perché ci interessa

Il **teorema dei valori intermedi**, incontrato informalmente nei primi moduli e dimostrato rigorosamente nel notebook 10.12, dice che una funzione continua su un intervallo $[a,b]$ assume tutti i valori tra $f(a)$ e $f(b)$. La dimostrazione standard usa l'assioma di completezza (notebook 10.9) e sfrutta un fatto geometrico intuitivo ma mai definito con precisione: un intervallo è "fatto di un unico pezzo", non si può spezzare in due parti separate. La **connessione** è la nozione topologica che cattura esattamente questa idea, permettendo di dimostrare il teorema dei valori intermedi (e le sue generalizzazioni) come conseguenza di un fatto puramente topologico, applicabile a spazi molto più generali di $\mathbb R$.

## Teoria, passo dopo passo

### Definizione di spazio connesso

Uno spazio topologico $X$ è **sconnesso** se esistono due aperti non vuoti $U,V$ con $U\cap V=\emptyset$ e $U\cup V=X$ (una "separazione" di $X$ in due pezzi aperti disgiunti). $X$ è **connesso** se non è sconnesso, cioè se gli unici sottoinsiemi di $X$ che sono **sia aperti sia chiusi** sono $\emptyset$ e $X$ stesso.

**Perché queste due formulazioni coincidono:** se $U,V$ separano $X$ (aperti, disgiunti, unione $X$), allora $V=X\setminus U$, quindi $U$ è anche chiuso (complementare dell'aperto $V$); viceversa, se $A$ è aperto e chiuso con $A\ne\emptyset,X$, allora $U=A$ e $V=X\setminus A$ (anch'esso aperto, essendo $A$ chiuso) separano $X$.

**Esempio di spazio sconnesso:** $X=[0,1]\cup[2,3]\subseteq\mathbb R$ (con la topologia di sottospazio, ereditata da $\mathbb R$: un insieme è aperto in $X$ se è intersezione di $X$ con un aperto di $\mathbb R$). Gli insiemi $U=[0,1]$ e $V=[2,3]$ sono entrambi aperti *in $X$* (anche se non in $\mathbb R$: per esempio $U=X\cap(-1,1.5)$), disgiunti, con unione $X$: $X$ è sconnesso.

### $\mathbb R$ è connesso

**Teorema.** $\mathbb R$ con la topologia euclidea è connesso.

**Dimostrazione (per assurdo, usando la completezza).** Supponiamo che $\mathbb R=U\cup V$ con $U,V$ aperti, disgiunti, non vuoti. Prendiamo $a\in U$, $b\in V$, e supponiamo $a<b$ (senza perdita di generalità). Consideriamo $s=\sup\{x\in[a,b]:x\in U\}$ (esiste per l'assioma di completezza, notebook 10.9, essendo l'insieme non vuoto — contiene $a$ — e limitato superiormente da $b$). Mostriamo che $s$ non può stare né in $U$ né in $V$, assurdo essendo $U\cup V=\mathbb R\ni s$.

*Se $s\in U$*: essendo $U$ aperto, esiste $\delta>0$ con $(s-\delta,s+\delta)\subseteq U$. Essendo $s\le b$ e $b\in V$ (disgiunto da $U$), deve essere $s<b$, quindi possiamo trovare punti $s<s+\delta'<b$ (con $\delta'<\delta$) che stanno ancora in $U$ (per l'aperto attorno a $s$) — questo contraddice la definizione di $s$ come **estremo superiore** dell'insieme dei punti di $U$ in $[a,b]$: avremmo trovato un punto di $U$ più grande di $s$.

*Se $s\in V$*: essendo $V$ aperto, esiste $\delta>0$ con $(s-\delta,s+\delta)\subseteq V$. Ma per definizione di estremo superiore, esistono punti di $U$ arbitrariamente vicini a $s$ da sinistra (altrimenti $s-\delta$ sarebbe un maggiorante più piccolo), quindi punti di $U$ dentro $(s-\delta,s+\delta)\subseteq V$ — assurdo, essendo $U,V$ disgiunti.

In entrambi i casi si trova un assurdo: $\mathbb R$ non può essere sconnesso. $\blacksquare$ (Questa dimostrazione è, in sostanza, la stessa tecnica usata per la completezza degli intervalli annidati nel notebook 10.9: la completezza di $\mathbb R$ è ciò che impedisce di "spezzarlo".)

### Continuità e connessione: generalizzazione del teorema dei valori intermedi

**Teorema.** Se $f:X\to Y$ è continua e $X$ è connesso, allora $f(X)$ (l'immagine, con la topologia di sottospazio di $Y$) è connesso.

**Dimostrazione.** Supponiamo per assurdo che $f(X)$ sia sconnesso: esistono aperti $U',V'$ (di $f(X)$) disgiunti, non vuoti, con unione $f(X)$. Allora $U=f^{-1}(U')$ e $V=f^{-1}(V')$ sono aperti in $X$ (per continuità di $f$, notebook 12.4), disgiunti (se $x\in U\cap V$, allora $f(x)\in U'\cap V'=\emptyset$, assurdo), non vuoti (essendo $U',V'$ non vuoti e $U',V'\subseteq f(X)$, hanno controimmagini non vuote), e con unione $X$ (ogni $x\in X$ ha $f(x)\in f(X)=U'\cup V'$, quindi $x\in U\cup V$). Questo separerebbe $X$, contraddicendo la sua connessione. $\blacksquare$

**Corollario (teorema dei valori intermedi, versione topologica).** Se $f:[a,b]\to\mathbb R$ è continua, $f([a,b])$ è connesso; si dimostra a parte (esercizio) che i sottoinsiemi connessi di $\mathbb R$ sono esattamente gli **intervalli** — dunque $f([a,b])$ è un intervallo, cioè contiene tutti i valori compresi tra $f(a)$ e $f(b)$: esattamente il teorema dei valori intermedi, ora ottenuto come caso particolare di un fatto puramente topologico.

```python
import numpy as np
import matplotlib.pyplot as plt

# Visualizziamo la dimostrazione: s = sup dei punti di U in [a,b] non puo' stare ne' in U ne' in V
fig, ax = plt.subplots(figsize=(8, 2.5))

a, b, s = 0, 5, 2.7
ax.axhline(0, color='gray', linewidth=0.5)
ax.plot([a, s], [0, 0], 'b-', linewidth=4, label='U (fino a s)')
ax.plot([s, b], [0, 0], 'r-', linewidth=4, label='V (da s)')
ax.plot(s, 0, 'ko', markersize=10, label='s = sup{x in [a,b] : x in U}')
ax.annotate('s deve stare in U o in V:\nentrambi i casi portano ad assurdo', xy=(s, 0),
            xytext=(s, 0.3), ha='center', fontsize=9)

ax.set_yticks([])
ax.set_xlim(a - 0.5, b + 0.5)
ax.set_ylim(-0.3, 0.6)
ax.legend(loc='lower right')
ax.set_title("Dimostrazione per assurdo: R e' connesso")
plt.tight_layout()
plt.show()

# Verifica numerica indiretta: proviamo a "separare" [0,5] con due aperti e vediamo che falliscono sempre
# vicino al punto di sup, restera' sempre un punto ambiguo
print("Non esiste una separazione valida di [0,5] in due aperti disgiunti non vuoti: R e i suoi intervalli sono connessi.")
```

## Esercizio svolto e commentato

**Problema:** dimostra che i sottoinsiemi connessi di $\mathbb R$ (con la topologia euclidea) sono esattamente gli **intervalli** (inclusi i casi degeneri: un punto, $\emptyset$, semirette, tutto $\mathbb R$).

**Ragionamento:**

**Passo 1 — ogni insieme connesso è un intervallo.** Dimostriamo il contronominale: se $A\subseteq\mathbb R$ **non** è un intervallo, allora $A$ è sconnesso. Non essere un intervallo significa che esistono $a,b\in A$ e $c\notin A$ con $a<c<b$ (altrimenti $A$ conterrebbe ogni punto tra due suoi punti, cioè sarebbe un intervallo per definizione). Allora $U=A\cap(-\infty,c)$ e $V=A\cap(c,+\infty)$ sono aperti *in $A$* (intersezioni di $A$ con aperti di $\mathbb R$), disgiunti (per costruzione, essendo separati da $c$), non vuoti ($a\in U$, $b\in V$), e la loro unione è $A$ (perché $c\notin A$, ogni punto di $A$ è o $<c$ o $>c$): $A$ è sconnesso.

**Passo 2 — ogni intervallo è connesso.** Sia $I$ un intervallo, e supponiamo per assurdo $I=U\cup V$ con $U,V$ aperti in $I$, disgiunti, non vuoti. Prendendo $a\in U$, $b\in V$ (con $a<b$ senza perdita di generalità — e $[a,b]\subseteq I$ perché $I$ è un intervallo), lo stesso argomento della dimostrazione di connessione di $\mathbb R$ nella teoria (usando $s=\sup\{x\in[a,b]:x\in U\}$, che esiste per completezza ed è in $[a,b]\subseteq I$) porta a un assurdo identico: $s$ non può stare né in $U$ né in $V$. Dunque nessuna separazione esiste: $I$ è connesso.

**Conclusione:** i due passi, insieme, danno l'equivalenza: $A\subseteq\mathbb R$ è connesso $\iff$ $A$ è un intervallo. **Punto metodologico:** questo esercizio mostra che la dimostrazione di connessione di $\mathbb R$ (teoria) non era un caso isolato, ma il prototipo di un argomento che funziona per **ogni** intervallo, semplicemente restringendo l'estremo superiore all'intervallo dato invece che a tutto $\mathbb R$.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 1, figsize=(8, 3.5))

# A = [0,1] U [2,3], NON un intervallo -> sconnesso
axes[0].plot([0, 1], [0, 0], 'b-', linewidth=4)
axes[0].plot([2, 3], [0, 0], 'r-', linewidth=4)
axes[0].axvline(1.5, color='gray', linestyle=':', label='c=1.5, punto mancante che separa')
axes[0].set_title("A = [0,1] U [2,3]: NON e' un intervallo -> sconnesso")
axes[0].legend()
axes[0].set_yticks([])

# I = [0,3], intervallo -> connesso
axes[1].plot([0, 3], [0, 0], 'g-', linewidth=4)
axes[1].set_title("I = [0,3]: intervallo -> connesso (nessun punto separatore esiste)")
axes[1].set_yticks([])

plt.tight_layout()
plt.show()

print("A = [0,1] U [2,3] e' sconnesso: separabile con U=A∩(-inf,1.5), V=A∩(1.5,+inf)")
print("I = [0,3] e' connesso: nessuna scelta di c in (0,3) puo' separarlo, essendo c stesso in I")
```

**1.** $f$ continua e $X$ connesso implicano, per il teorema della teoria, che $f(X)$ è connesso. Essendo $f$ suriettiva, $f(X)=Y$: dunque $Y$ è connesso.

**2.** Consideriamo $f:[a,b]\to\mathbb R$ continua con $f(a)<0<f(b)$. $[a,b]$ è un intervallo, quindi connesso (esercizio svolto). Per il teorema della teoria, $f([a,b])$ è connesso, quindi (essendo un sottoinsieme di $\mathbb R$) è un intervallo (esercizio svolto, l'altra direzione). Nella topologia discreta, **ogni** sottoinsieme è aperto (per definizione), quindi sia $U$ sia $V$ sono aperti; sono disgiunti per costruzione, non vuoti ($U$ contiene $x$, $V$ contiene almeno un altro punto essendo $|X|\ge2$), e la loro unione è $X$. Questa è una separazione: $X$ è sconnesso.

**4.** Supponiamo per assurdo $A\cup B=U\cup V$ (aperti in $A\cup B$, disgiunti, non vuoti). Sia $p\in A\cap B$ (esiste per ipotesi); $p$ sta in $U$ oppure in $V$ (non entrambi, essendo disgiunti) — supponiamo $p\in U$ senza perdita di generalità. Consideriamo $U\cap A$ e $V\cap A$: sono aperti in $A$ (intersezioni di aperti di $A\cup B$ con $A$), disgiunti, e la loro unione è $A$ (essendo $A\subseteq A\cup B=U\cup V$). Essendo $A$ connesso, uno dei due deve essere vuoto: essendo $p\in U\cap A$ (perché $p\in A$ e $p\in U$) non vuoto, deve essere $V\cap A=\emptyset$, cioè $A\subseteq U$. Con lo stesso argomento su $B$ (che contiene anch'esso $p\in U$): $B\cap V=\emptyset$, cioè $B\subseteq U$. Dunque $A\cup B\subseteq U$, cioè $V=\emptyset$ — contraddicendo l'ipotesi che $V$ fosse non vuoto. Dunque nessuna separazione esiste: $A\cup B$ è connesso.

## Riepilogo finale

- Uno spazio è **connesso** se non si può separare in due aperti disgiunti non vuoti, equivalentemente se gli unici insiemi sia aperti sia chiusi sono $\emptyset$ e $X$.
- $\mathbb R$ è connesso: la dimostrazione usa l'assioma di completezza (notebook 10.9), lo stesso strumento cruciale che regge tutta l'analisi reale rigorosa.
- La connessione è preservata da funzioni continue: $f$ continua e $X$ connesso $\Rightarrow$ $f(X)$ connesso — da qui segue il teorema dei valori intermedi come caso particolare.
- I sottoinsiemi connessi di $\mathbb R$ sono esattamente gli **intervalli**: questa caratterizzazione traduce l'intuizione geometrica ("un solo pezzo") in un fatto dimostrabile con precisione.
