# 12.12 — Gruppo fondamentale

**Prerequisiti:** 12.11 Omeomorfismi, 11.1 Teoria dei gruppi, 11.4 Omomorfismi, 12.5 Connessione.


## Perché ci interessa

Nel notebook 12.11 abbiamo dimostrato che $\mathbb R\not\cong\mathbb R^2$ togliendo un punto e guardando la connessione risultante — una tecnica ad hoc, che ha funzionato ma non si generalizza facilmente (per esempio, non distingue subito $\mathbb R^2$ da $\mathbb R^3$, dove togliere un punto lascia entrambi connessi). Serve un invariante più fine, capace di "sentire" i buchi di uno spazio in modo sistematico. Il **gruppo fondamentale**, l'idea più famosa della topologia algebrica, associa a ogni spazio (con un punto base) un gruppo — usando esattamente le strutture del notebook 11.1 — che misura in quanti modi essenzialmente diversi un cappio può essere disegnato nello spazio. Un disco non ha buchi: ogni cappio si può contrarre a un punto. Una circonferenza (o un disco bucato) ha un buco: un cappio che gira attorno al buco non si può contrarre senza attraversarlo.

## Teoria, passo dopo passo

### Cammini e omotopia

Un **cammino** in $X$ da $x_0$ a $x_1$ è una funzione continua $\gamma:[0,1]\to X$ con $\gamma(0)=x_0$, $\gamma(1)=x_1$. Un **cappio** basato in $x_0$ è un cammino con $\gamma(0)=\gamma(1)=x_0$.

Due cappi $\gamma_0,\gamma_1$ basati in $x_0$ sono **omotopi** (scritto $\gamma_0\simeq\gamma_1$) se esiste una funzione continua $H:[0,1]\times[0,1]\to X$ (una "deformazione continua", usando la topologia prodotto del notebook 12.9) tale che $H(s,0)=\gamma_0(s)$, $H(s,1)=\gamma_1(s)$, e $H(0,t)=H(1,t)=x_0$ per ogni $t$ (il punto base resta fisso durante tutta la deformazione). Intuitivamente, $H(\cdot,t)$ è il cappio "a metà cammino" tra $\gamma_0$ e $\gamma_1$, al variare di $t$ da $0$ a $1$.

**L'omotopia è una relazione di equivalenza** sui cappi basati in $x_0$: riflessiva ($H(s,t)=\gamma(s)$, costante in $t$), simmetrica ($H'(s,t)=H(s,1-t)$), transitiva (concatenando due omotopie, riscalando il parametro $t$ — verifica tecnica di continuità che omettiamo, analoga a quella per la composizione di cammini sotto).

### Concatenazione di cammini e il gruppo fondamentale

Dati due cappi $\gamma,\delta$ basati in $x_0$, la loro **concatenazione** $\gamma*\delta$ è il cappio che percorre $\gamma$ nella prima metà del tempo e $\delta$ nella seconda: $(\gamma*\delta)(s)=\gamma(2s)$ per $s\in[0,1/2]$, $(\gamma*\delta)(s)=\delta(2s-1)$ per $s\in[1/2,1]$ (ben definita e continua: i due pezzi coincidono in $s=1/2$, dando $\gamma(1)=x_0=\delta(0)$).

Il **gruppo fondamentale** $\pi_1(X,x_0)$ è l'insieme delle classi di omotopia $[\gamma]$ dei cappi basati in $x_0$, con operazione $[\gamma]*[\delta]=[\gamma*\delta]$ (ben definita: si dimostra che l'omotopia è compatibile con la concatenazione, in modo simile alla verifica di buona posizione del notebook 11.1 per $\mathbb Z/n\mathbb Z$).

**Teorema (enunciato, verifica omessa).** $(\pi_1(X,x_0),*)$ è un gruppo: l'elemento neutro è la classe del cappio **costante** $e(s)=x_0$; l'inverso di $[\gamma]$ è $[\gamma^{-1}]$, dove $\gamma^{-1}(s)=\gamma(1-s)$ (lo stesso cammino percorso al contrario); l'associatività e le proprietà del neutro/inverso si verificano costruendo esplicitamente le omotopie necessarie (per esempio, $\gamma*\gamma^{-1}\simeq e$ tramite una deformazione che "arretra" progressivamente il punto di inversione). La struttura di gruppo qui è **esattamente** quella degli assiomi del notebook 11.1, applicata a un insieme di classi di cammini invece che a numeri o matrici.

### Il gruppo fondamentale del disco è banale

**Teorema.** $\pi_1(D,x_0)=\{e\}$ per $D=\{(x,y):x^2+y^2\le1\}$ (disco chiuso), qualunque sia $x_0\in D$.

**Dimostrazione.** Dato un cappio $\gamma$ basato in $x_0$, definiamo $H(s,t)=(1-t)\gamma(s)+t\,x_0$ (interpolazione lineare tra $\gamma(s)$ e il punto fisso $x_0$ — ben definita, perché $D$ è **convesso**: il segmento tra due suoi punti resta in $D$). $H$ è continua (composizione di operazioni continue: moltiplicazione per scalare, somma, entrambe continue in $\mathbb R^2$), $H(s,0)=\gamma(s)$, $H(s,1)=x_0$ (il cappio costante), e $H(0,t)=H(1,t)=(1-t)x_0+t\,x_0=x_0$ (punto base fisso, essendo $\gamma(0)=\gamma(1)=x_0$). Dunque ogni cappio è omotopo al cappio costante: $\pi_1(D,x_0)$ ha un solo elemento. $\blacksquare$ Uno spazio con $\pi_1$ banale si dice **semplicemente connesso**: intuitivamente, "senza buchi che intrappolano i cappi".

### Il gruppo fondamentale della circonferenza (enunciato)

**Teorema (enunciato, dimostrazione fuori portata di questo notebook introduttivo).** $\pi_1(S^1,x_0)\cong(\mathbb Z,+)$: la classe di un cappio è determinata univocamente dal suo **numero di avvolgimento** (quante volte, e in che verso, il cappio gira attorno al centro prima di richiudersi). Il cappio costante ha avvolgimento $0$; percorrere $S^1$ una volta in senso antiorario dà avvolgimento $1$; concatenare due giri dà avvolgimento $2$ (coerente con l'operazione di gruppo, che corrisponde alla somma degli interi, notebook 11.1); percorrere in senso orario dà avvolgimento $-1$ (l'inverso del gruppo). La dimostrazione rigorosa richiede la teoria dei **rivestimenti** (notebook 12.13), che fornisce lo strumento tecnico (il sollevamento dei cammini) per definire con precisione il numero di avvolgimento.

### Invarianza per omotopia e conseguenza sugli omeomorfismi

**Teorema (enunciato).** Se $X\cong Y$ tramite un omeomorfismo $f$ con $f(x_0)=y_0$, allora $\pi_1(X,x_0)\cong\pi_1(Y,y_0)$ (isomorfismo indotto da $\gamma\mapsto f\circ\gamma$). Dunque $\pi_1$ è un **invariante topologico**, come compattezza e connessione (notebook 12.11), ma molto più sensibile: essendo $\pi_1(D,x_0)=\{e\}\ne\mathbb Z\cong\pi_1(S^1,x_0)$ (gruppi non isomorfi — uno banale, l'altro infinito), $D\not\cong S^1$, un fatto che né la compattezza né la connessione (identiche per entrambi) potevano rilevare.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

# Nel disco, ogni cappio si contrae al centro (omotopia)
theta = np.linspace(0, 2*np.pi, 100)
for t in [0, 0.3, 0.6, 1.0]:
    raggio = 0.8 * (1 - t)
    axes[0].plot(raggio*np.cos(theta), raggio*np.sin(theta), alpha=0.5 + 0.5*t,
                 label=f"H(.,{t}) " + ("(cappio originale)" if t==0 else "(contratto)" if t==1 else ""))
axes[0].plot(0, 0, 'ko', markersize=8, label="punto base x0 = centro")
axes[0].set_aspect('equal')
axes[0].set_title("Disco: ogni cappio si contrae -> pi_1(D) banale")
axes[0].legend(fontsize=7, loc='upper right')

# Su S^1, un cappio che gira una volta NON puo' contrarsi (dovrebbe attraversare il "buco" al centro)
axes[1].plot(np.cos(theta), np.sin(theta), 'g-', linewidth=3, label="cappio con avvolgimento 1")
axes[1].plot(0, 0, 'rx', markersize=15, label="buco (punto assente)")
axes[1].set_aspect('equal')
axes[1].set_title("S^1 stesso: il cappio non puo' contrarsi restando su S^1")
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.show()

print("pi_1(disco) = {e} (banale, semplicemente connesso)")
print("pi_1(S^1) = Z (numero di avvolgimento) -> disco e circonferenza NON sono omeomorfi")
```

## Esercizio svolto e commentato

**Problema:** dimostra che se $X$ è **convesso** in $\mathbb R^n$ (per ogni $x,y\in X$, il segmento $\{(1-t)x+ty:t\in[0,1]\}\subseteq X$), allora $X$ è semplicemente connesso (generalizzando la dimostrazione per il disco vista nella teoria).

**Ragionamento:**

**Passo 1 — generalizziamo la costruzione della teoria.** Sia $\gamma$ un cappio in $X$ basato in $x_0$. Definiamo $H(s,t)=(1-t)\gamma(s)+t\,x_0$, esattamente come nella dimostrazione per il disco.

**Passo 2 — verifichiamo che $H$ è ben definita in $X$.** Per ogni $s,t$ fissati, $H(s,t)$ è un punto sul segmento tra $\gamma(s)\in X$ e $x_0\in X$; per convessità di $X$, questo segmento è interamente contenuto in $X$, quindi $H(s,t)\in X$. **Questo è l'unico punto dove serve la convessità** — è precisamente ciò che manca a $S^1$ (non convesso: il segmento tra due punti della circonferenza, tranne casi degeneri, esce dalla circonferenza) e che rende la dimostrazione inapplicabile lì.

**Passo 3 — verifichiamo continuità e condizioni al bordo.** $H$ è continua, essendo composizione di somma e moltiplicazione per scalare (continue in $\mathbb R^n$) con $\gamma$ (continua per ipotesi) — formalmente, $H$ è continua come funzione dal prodotto $[0,1]\times[0,1]$ (notebook 12.9) a $\mathbb R^n$, e la sua immagine sta in $X$ per il passo 2, quindi è continua anche come funzione a valori in $X$. $H(s,0)=\gamma(s)$, $H(s,1)=x_0$, $H(0,t)=H(1,t)=x_0$ (usando $\gamma(0)=\gamma(1)=x_0$): tutte le condizioni di omotopia sono soddisfatte.

**Conclusione:** ogni cappio in $X$ è omotopo al cappio costante: $\pi_1(X,x_0)=\{e\}$, cioè $X$ è semplicemente connesso. **Portata del risultato:** questo dimostra in un colpo solo che $\mathbb R^n$ stesso, ogni palla, ogni semispazio, ogni insieme convesso qualsiasi (una classe molto ampia di spazi, incontrata fin dai notebook 6.1-6.2 sui vettori e la geometria vettoriale) è semplicemente connesso — nessuno di questi spazi può essere omeomorfo a $S^1$ o a qualunque spazio con un "buco".

```python
import numpy as np
import matplotlib.pyplot as plt

# Verifichiamo la convessita' del quadrato [0,1]x[0,1] e la non-convessita' di S^1
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Quadrato: convesso
quad_x, quad_y = [0,1,1,0,0], [0,0,1,1,0]
axes[0].fill(quad_x, quad_y, alpha=0.3, color='blue')
p1, p2 = np.array([0.1, 0.1]), np.array([0.9, 0.8])
segmento = np.linspace(0, 1, 20)
seg_points = np.outer(1-segmento, p1) + np.outer(segmento, p2)
axes[0].plot(seg_points[:,0], seg_points[:,1], 'ro-', markersize=3)
axes[0].set_title("Quadrato: CONVESSO\n(ogni segmento tra due punti resta dentro)")
axes[0].set_aspect('equal')

# Circonferenza: non convessa
theta = np.linspace(0, 2*np.pi, 100)
axes[1].plot(np.cos(theta), np.sin(theta), 'g-', linewidth=2)
q1, q2 = np.array([1, 0]), np.array([-1, 0])
seg_points2 = np.outer(1-segmento, q1) + np.outer(segmento, q2)
axes[1].plot(seg_points2[:,0], seg_points2[:,1], 'ro-', markersize=3, label='segmento ESCE da S^1')
axes[1].set_aspect('equal')
axes[1].set_title("Circonferenza: NON convessa")
axes[1].legend()

plt.tight_layout()
plt.show()

print("Il quadrato e' convesso -> semplicemente connesso (per l'esercizio svolto)")
print("La circonferenza non e' convessa -> l'argomento non si applica (e infatti pi_1(S^1)=Z, non banale)")
```

## Riepilogo finale

- Il **gruppo fondamentale** $\pi_1(X,x_0)$ raccoglie le classi di omotopia dei cappi basati in $x_0$, con la concatenazione come operazione — una struttura di gruppo (notebook 11.1) costruita su oggetti geometrici invece che numerici.
- Spazi **convessi** (dischi, $\mathbb R^n$, ogni insieme convesso) hanno gruppo fondamentale banale: **semplicemente connessi**, ogni cappio si contrae a un punto.
- $\pi_1(S^1)\cong\mathbb Z$ (numero di avvolgimento): un invariante molto più fine di compattezza e connessione, capace di distinguere spazi che quelle proprietà non distinguono.
- $\pi_1$ è invariante per omeomorfismo: due spazi con gruppi fondamentali non isomorfi non possono essere omeomorfi, la tecnica più potente vista finora per dimostrare la non-equivalenza topologica.
