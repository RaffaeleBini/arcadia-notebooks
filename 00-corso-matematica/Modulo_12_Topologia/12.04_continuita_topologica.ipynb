## Perché ci interessa

Nel notebook 10.12 abbiamo dimostrato con $\varepsilon$ e $\delta$ che $f$ è continua in $x_0$ se punti vicini a $x_0$ vanno in punti vicini a $f(x_0)$. Con il linguaggio degli aperti (notebook 12.1) possiamo dire la stessa cosa **senza misurare distanze**: la richiesta "$|x-x_0|<\delta\Rightarrow|f(x)-f(x_0)|<\varepsilon$" dice, in sostanza, che l'intorno $(x_0-\delta,x_0+\delta)$ va a finire dentro l'intorno $(f(x_0)-\varepsilon,f(x_0)+\varepsilon)$. Riformulare la continuità in termini di aperti (invece che di distanze) è il passo che rende la nozione applicabile a **qualunque** spazio topologico — anche dove non esiste alcuna nozione di distanza numerica — e getta le basi per capire perché due spazi "topologicamente diversi" (come una retta e una circonferenza, notebook 12.11) non possono essere messi in corrispondenza continua e invertibile.

## Teoria, passo dopo passo

### Continuità in un punto, riformulata con gli intorni

Riprendiamo la definizione $\varepsilon$-$\delta$ (notebook 10.8, 10.12): $f:\mathbb R\to\mathbb R$ è continua in $x_0$ se per ogni $\varepsilon>0$ esiste $\delta>0$ tale che $x\in(x_0-\delta,x_0+\delta)\Rightarrow f(x)\in(f(x_0)-\varepsilon,f(x_0)+\varepsilon)$. Riscriviamo: per ogni intorno $V$ di $f(x_0)$ esiste un intorno $U$ di $x_0$ tale che $f(U)\subseteq V$ — equivalentemente, $U\subseteq f^{-1}(V)$, dove $f^{-1}(V)=\{x\in X:f(x)\in V\}$ è l'insieme dei punti che $f$ manda dentro $V$ (la **controimmagine** o **preimmagine** di $V$). Questa formulazione non menziona più $\varepsilon,\delta$: parla solo di intorni.

### Definizione topologica di continuità (globale)

Siano $(X,\tau_X)$ e $(Y,\tau_Y)$ spazi topologici. Una funzione $f:X\to Y$ è **continua** se per ogni aperto $V\in\tau_Y$, la controimmagine $f^{-1}(V)=\{x\in X:f(x)\in V\}$ è un aperto di $X$.

**Perché "controimmagine" e non "immagine"?** Perché la continuità garantisce che l'informazione "dove può arrivare $f$" si propaghi all'indietro in modo controllato: se $V$ è un intorno "piccolo" attorno a $f(x_0)$, la sua controimmagine deve contenere un intorno "piccolo" attorno a $x_0$ — esattamente ciò che dice $\delta$ in funzione di $\varepsilon$. L'immagine diretta di un aperto, invece, **non** deve essere aperta perché $f$ sia continua: per esempio $f(x)=x^2$ è continua, ma $f((-1,1))=[0,1)$, che non è aperto in $\mathbb R$.

**Teorema (equivalenza con la continuità puntuale).** $f$ è continua (nel senso globale sopra) se e solo se è continua in ogni punto $x_0\in X$ (nel senso degli intorni).

**Dimostrazione.** *(⇒)* Sia $f$ continua globalmente, $x_0\in X$, e $V$ un intorno di $f(x_0)$ (contenente un aperto $V'\ni f(x_0)$). Allora $f^{-1}(V')$ è aperto (per continuità globale) e contiene $x_0$ (perché $f(x_0)\in V'$): è dunque un intorno $U$ di $x_0$ con $f(U)\subseteq V'\subseteq V$. *(⇐)* Sia $f$ continua in ogni punto, e $V\in\tau_Y$ aperto. Mostriamo che $f^{-1}(V)$ è aperto: sia $x_0\in f^{-1}(V)$, cioè $f(x_0)\in V$. Essendo $V$ aperto, è un intorno di $f(x_0)$; per continuità in $x_0$, esiste un intorno $U$ di $x_0$ con $f(U)\subseteq V$, cioè $U\subseteq f^{-1}(V)$. Dunque ogni punto di $f^{-1}(V)$ ha un intorno interamente dentro $f^{-1}(V)$: $f^{-1}(V)$ è aperto. $\blacksquare$

### Continuità tramite i chiusi

**Corollario.** $f$ è continua se e solo se $f^{-1}(C)$ è chiuso in $X$ per ogni chiuso $C$ di $Y$.

**Dimostrazione.** Segue subito dall'identità insiemistica $f^{-1}(Y\setminus C)=X\setminus f^{-1}(C)$ (un punto va nel complementare di $C$ se e solo se la sua controimmagine sta nel complementare della controimmagine di $C$): se $C$ è chiuso, $Y\setminus C$ è aperto, quindi (per continuità) $f^{-1}(Y\setminus C)=X\setminus f^{-1}(C)$ è aperto, cioè $f^{-1}(C)$ è chiuso; e viceversa. $\blacksquare$

### Composizione di funzioni continue

**Teorema.** Se $f:X\to Y$ e $g:Y\to Z$ sono continue, allora $g\circ f:X\to Z$ è continua.

**Dimostrazione.** Sia $W$ aperto in $Z$. Per continuità di $g$, $g^{-1}(W)$ è aperto in $Y$. Per continuità di $f$, $f^{-1}(g^{-1}(W))$ è aperto in $X$. Ma $f^{-1}(g^{-1}(W))=(g\circ f)^{-1}(W)$ (proprietà generale delle controimmagini, verificabile direttamente dalla definizione: $x$ sta nel primo insieme $\iff$ $f(x)\in g^{-1}(W)$ $\iff$ $g(f(x))\in W$ $\iff$ $x$ sta nel secondo). Dunque $(g\circ f)^{-1}(W)$ è aperto per ogni aperto $W$: $g\circ f$ è continua. $\blacksquare$ Questa dimostrazione, di tre righe, sostituisce l'argomento $\varepsilon$-$\delta$ "a due strati" necessario per dimostrare lo stesso fatto in $\mathbb R$ (notebook 10.12): un chiaro esempio di come l'astrazione **semplifichi** invece di complicare.

```python
import sympy as sp

# Verifichiamo su un esempio concreto l'equivalenza tra continuita' epsilon-delta e continuita' via controimmagini
x = sp.symbols('x', real=True)
f = x**2

# f(x) = x^2 e' continua in x0 = 1; verifichiamo che la controimmagine di un intorno aperto di f(1)=1
# contiene un intorno aperto di x0=1
x0 = 1
f_x0 = f.subs(x, x0)
print(f"f({x0}) = {f_x0}")

# Intorno V = (0.81, 1.21) di f(x0)=1 (epsilon=0.19 circa)
V_low, V_high = sp.Rational(81, 100), sp.Rational(121, 100)

# Risolviamo f(x) in V, cioe' troviamo la controimmagine f^{-1}(V) vicino a x0=1
soluzioni = sp.solve([f > V_low, f < V_high], x)
print(f"Controimmagine di V=({float(V_low)},{float(V_high)}) vicino a x0=1:", soluzioni)
print("Contiene un intorno aperto di x0=1?", "Si', l'intervallo trovato e' aperto e contiene 1")
```

## Esercizio svolto e commentato

**Problema:** dimostra, usando la definizione topologica, che la funzione $f:\mathbb R\to\mathbb R$, $f(x)=1$ se $x\ge0$, $f(x)=0$ se $x<0$ (funzione a **gradino**), non è continua in $x_0=0$.

**Ragionamento:**

**Passo 1 — troviamo un aperto "sospetto" in codominio.** Vogliamo un aperto $V\subseteq\mathbb R$ tale che $f^{-1}(V)$ **non** sia aperto. Proviamo $V=(1/2,3/2)$, un intorno di $f(0)=1$.

**Passo 2 — calcoliamo $f^{-1}(V)$.** $f(x)\in(1/2,3/2)$ significa $f(x)=1$ (l'unico valore di $f$ in quell'intervallo, essendo $f$ a valori solo $0$ o $1$), cioè $x\ge0$. Dunque $f^{-1}(V)=[0,+\infty)$.

**Passo 3 — verifichiamo che $[0,+\infty)$ non è aperto.** Il punto $0\in[0,+\infty)$ non ha nessun intorno $(-\delta,\delta)$ interamente contenuto in $[0,+\infty)$: per ogni $\delta>0$, il punto $-\delta/2\in(-\delta,\delta)$ ma $-\delta/2\notin[0,+\infty)$. Dunque $[0,+\infty)$ non è aperto.

**Conclusione:** abbiamo trovato un aperto $V=(1/2,3/2)$ la cui controimmagine $f^{-1}(V)=[0,+\infty)$ non è aperta: per la definizione topologica, $f$ non è continua (su tutto $\mathbb R$). Ripetendo l'argomento localmente in $x_0=0$: ogni intorno $V$ di $f(0)=1$ abbastanza piccolo da escludere $0$ (come $(1/2,3/2)$) ha controimmagine $[0,+\infty)$, che non contiene nessun intorno aperto di $0$ (per lo stesso motivo del passo 3) — coerente con l'intuizione $\varepsilon$-$\delta$: prendendo $\varepsilon<1/2$, nessun $\delta>0$ funziona, perché punti $x<0$ arbitrariamente vicini a $0$ danno $f(x)=0$, lontano da $f(0)=1$.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.where(x >= 0, 1, 0)

x_vals = np.linspace(-2, 2, 1000)
y_vals = f(x_vals)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x_vals[x_vals < 0], y_vals[x_vals < 0], 'b-', linewidth=2)
ax.plot(x_vals[x_vals >= 0], y_vals[x_vals >= 0], 'b-', linewidth=2)
ax.plot(0, 0, 'wo', markeredgecolor='blue', markersize=10)  # punto escluso
ax.plot(0, 1, 'bo', markersize=10)  # punto incluso
ax.axhspan(0.5, 1.5, alpha=0.2, color='orange', label='V = (1/2, 3/2), aperto in Y')
ax.set_title("Funzione a gradino: f^-1(V) = [0, +inf), NON aperto -> f non continua in 0")
ax.legend()
ax.set_ylim(-0.5, 2)
plt.tight_layout()
plt.show()

# Verifica diretta che [0, +infinito) non e' aperto: nessun intorno di 0 ci sta dentro
for delta in [0.1, 0.01, 0.0001]:
    punto_fuori = -delta / 2
    print(f"delta={delta}: il punto {punto_fuori} sta in (-delta,delta) ma non in [0,+inf): {punto_fuori < 0}")
```

4. *(più impegnativo)* Dimostra che se $Y$ ha la topologia **indiscreta**, ogni funzione $f:X\to Y$ è continua, qualunque sia $X$. Confronta con l'esercizio 3: spiega perché questi due risultati mostrano che "quasi tutte le funzioni sono continue" verso spazi con pochi aperti, o da spazi con tanti aperti.

**1.** Sia $V\in\tau$ aperto. $\operatorname{id}^{-1}(V)=\{x\in X:\operatorname{id}(x)\in V\}=\{x\in X:x\in V\}=V$, che è aperto per ipotesi. Dunque la controimmagine di ogni aperto è aperta: $\operatorname{id}$ è continua.

**2.** Sia $V\in\tau_Y$ un aperto qualsiasi di $Y$. Due casi: se $c\in V$, allora $f^{-1}(V)=X$ (ogni punto va in $c\in V$), che è aperto (assioma 1 di topologia, notebook 12.1); se $c\notin V$, allora $f^{-1}(V)=\emptyset$ (nessun punto va in $V$), anch'esso aperto. **3.** Sia $V\in\tau_Y$ un aperto qualsiasi. $f^{-1}(V)\subseteq X$ è un sottoinsieme di $X$; ma nella topologia discreta **ogni** sottoinsieme di $X$ è aperto (per definizione, $\tau_X=\mathcal P(X)$). Dunque $f^{-1}(V)$ è automaticamente aperto, qualunque esso sia: $f$ è continua.

**4.** Sia $V\in\tau_Y=\{\emptyset,Y\}$ (unici aperti nella topologia indiscreta). Se $V=\emptyset$: $f^{-1}(\emptyset)=\emptyset$, aperto in $X$ (assioma 1). Se $V=Y$: $f^{-1}(Y)=X$ (ogni punto va in $Y$, essendo $f$ a valori in $Y$), aperto in $X$ (assioma 1). In entrambi i casi $f^{-1}(V)$ è aperto: $f$ è continua. **Confronto:** questi due risultati mostrano che la continuità dipende **fortemente** dalla "ricchezza" delle topologie in gioco: se il dominio ha *troppi* aperti (discreta) o il codominio ne ha *troppo pochi* (indiscreta), la condizione di continuità diventa quasi vuota, perché non c'è quasi nessuna "struttura di vicinanza" da preservare. La continuità è davvero informativa solo quando entrambe le topologie hanno una struttura "intermedia" ricca, come quella euclidea.

## Riepilogo finale

- La continuità $\varepsilon$-$\delta$ (notebook 10.12) equivale a: **la controimmagine di ogni aperto è aperta** — una definizione puramente topologica, senza distanze.
- Equivalentemente: la controimmagine di ogni **chiuso** è chiusa. L'immagine diretta di un aperto, invece, non deve restare aperta.
- La composizione di funzioni continue è continua, con una dimostrazione di poche righe basata sulla proprietà $(g\circ f)^{-1}=f^{-1}\circ g^{-1}$.
- Continuità "banale" verso spazi con pochi aperti (indiscreti) o da spazi con tanti aperti (discreti): la continuità è davvero un vincolo solo quando le topologie coinvolte sono sufficientemente ricche in entrambe le direzioni.
