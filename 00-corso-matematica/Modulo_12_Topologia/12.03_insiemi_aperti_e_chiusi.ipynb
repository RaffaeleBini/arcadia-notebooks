Ma nel linguaggio comune della matematica si parla anche di insiemi **chiusi** — come $[a,b]$ rispetto a $(a,b)$ — e di operazioni come "il più piccolo chiuso che contiene un insieme dato" (la sua **chiusura**). ## Teoria, passo dopo passo

### Insiemi chiusi

Un insieme $C\subseteq X$ è **chiuso** se il suo complementare $X\setminus C$ è aperto. Per le **leggi di De Morgan** (notebook 0.13) applicate agli assiomi di topologia (notebook 12.1), i chiusi soddisfano proprietà "duali" a quelle degli aperti:

1. $\emptyset$ e $X$ sono chiusi (sono complementari l'uno dell'altro, ed entrambi aperti);
2. l'**intersezione** di una famiglia qualsiasi di chiusi è chiusa (complementare di un'unione di aperti, che è aperta);
3. l'**unione di un numero finito** di chiusi è chiusa (complementare di un'intersezione finita di aperti).

**Attenzione**: "chiuso" **non** è l'opposto di "aperto" nel senso che un insieme può essere sia aperto sia chiuso (come $\emptyset$ e $X$ sempre), oppure né aperto né chiuso (come $[0,1)$ in $\mathbb R$ con la topologia euclidea: non è aperto perché $0$ non ha un intorno interamente dentro $[0,1)$; non è chiuso perché il suo complementare $(-\infty,0)\cup[1,+\infty)$ non è aperto, dato che $1$ non ha un intorno dentro il complementare).

### Punti di accumulazione e chiusura

Un punto $x\in X$ è **punto di accumulazione** (o punto limite) di $A\subseteq X$ se ogni intorno di $x$ contiene almeno un punto di $A$ diverso da $x$ stesso. La **chiusura** di $A$, indicata $\overline A$, è $A$ unito ai suoi punti di accumulazione.

**Teorema.** $\overline A$ è il più piccolo chiuso che contiene $A$ (cioè $\overline A=\bigcap\{C\text{ chiuso}:A\subseteq C\}$).

**Dimostrazione.** Mostriamo prima che $\overline A$ è chiuso, cioè che $X\setminus\overline A$ è aperto. Sia $y\notin\overline A$: allora $y\notin A$ e $y$ non è punto di accumulazione di $A$, quindi esiste un intorno $U$ di $y$ che non contiene punti di $A$ diversi da $y$ — ma $y\notin A$, quindi $U$ non contiene **nessun** punto di $A$, cioè $U\subseteq X\setminus A$. Inoltre, nessun punto di $U$ può essere di accumulazione per $A$ (se $z\in U$ fosse di accumulazione, $U$ stesso — intorno di $z$, essendo aperto — dovrebbe contenere punti di $A$, contraddizione), quindi $U\subseteq X\setminus\overline A$: ogni punto fuori da $\overline A$ ha un intorno interamente fuori da $\overline A$, cioè $X\setminus\overline A$ è aperto, cioè $\overline A$ è chiuso. Ora, sia $C$ un chiuso qualsiasi con $A\subseteq C$: se $x$ è di accumulazione per $A$, ogni intorno di $x$ contiene punti di $A\subseteq C$; se fosse $x\notin C$, allora $x\in X\setminus C$ (aperto), che sarebbe un intorno di $x$ senza punti di $C\supseteq A$ — contraddizione. Dunque $x\in C$: ogni punto di accumulazione di $A$ sta in $C$, cioè $\overline A\subseteq C$. Essendo $C$ arbitrario tra i chiusi contenenti $A$, $\overline A$ è il più piccolo di essi. $\blacksquare$

### Interno e frontiera

L'**interno** di $A$, $\operatorname{int}(A)$ (o $A^\circ$), è l'unione di tutti gli aperti contenuti in $A$: il più grande aperto dentro $A$ (per dualità con la chiusura, la dimostrazione è analoga a quella sopra, scambiando aperti/chiusi e unione/intersezione). La **frontiera** di $A$ è $\partial A=\overline A\setminus\operatorname{int}(A)$: i punti "al confine", che appartengono alla chiusura ma non all'interno.

**Esempio in $\mathbb R$:** per $A=(0,1]$, $\operatorname{int}(A)=(0,1)$ (il più grande aperto dentro $A$: $1$ non è interno, perché nessun intorno di $1$ sta interamente in $(0,1]$), $\overline A=[0,1]$ ($0$ è punto di accumulazione anche se $0\notin A$: ogni intorno di $0$ contiene punti di $(0,1]$), e $\partial A=[0,1]\setminus(0,1)=\{0,1\}$.

### Un insieme è chiuso se e solo se coincide con la propria chiusura

**Corollario.** $A$ è chiuso $\iff A=\overline A$ $\iff$ $A$ contiene tutti i suoi punti di accumulazione. Segue immediatamente dal teorema: se $A$ è chiuso, essendo esso stesso un chiuso che contiene $A$, per minimalità $\overline A\subseteq A$, e sempre $A\subseteq\overline A$, quindi $A=\overline A$. Viceversa, se $A=\overline A$, allora $A$ è chiuso perché $\overline A$ lo è sempre.

```python
import numpy as np
import matplotlib.pyplot as plt

# Visualizziamo interno, chiusura e frontiera di A = (0, 1] in R
fig, axes = plt.subplots(3, 1, figsize=(8, 4), sharex=True)

# A = (0, 1]
axes[0].plot([0, 1], [0, 0], 'b-', linewidth=3)
axes[0].plot(0, 0, 'wo', markeredgecolor='blue', markersize=8)  # 0 escluso
axes[0].plot(1, 0, 'bo', markersize=8)  # 1 incluso
axes[0].set_title("A = (0, 1]")

# interno = (0, 1)
axes[1].plot([0, 1], [0, 0], 'g-', linewidth=3)
axes[1].plot(0, 0, 'wo', markeredgecolor='green', markersize=8)
axes[1].plot(1, 0, 'wo', markeredgecolor='green', markersize=8)
axes[1].set_title("int(A) = (0, 1)")

# chiusura = [0, 1]
axes[2].plot([0, 1], [0, 0], 'r-', linewidth=3)
axes[2].plot(0, 0, 'ro', markersize=8)
axes[2].plot(1, 0, 'ro', markersize=8)
axes[2].set_title("chiusura(A) = [0, 1]  (frontiera = {0, 1})")

for ax in axes:
    ax.set_yticks([])
    ax.set_xlim(-0.3, 1.3)
plt.tight_layout()
plt.show()

# Verifica: 0 e' punto di accumulazione di A anche se 0 non appartiene ad A
A_campione = np.linspace(1e-6, 1, 100000)  # punti di A vicino a 0
print("Punto piu' vicino a 0 in A:", A_campione.min(), "-> per ogni intorno di 0, esistono punti di A dentro")
```

## Esercizio svolto e commentato

**Problema:** in $\mathbb R^2$ con la topologia euclidea, sia $A=\{(x,y):x^2+y^2<1\}\cup\{(1,0)\}$ (il disco aperto unitario, più un punto isolato sul bordo). Determina $\operatorname{int}(A)$, $\overline A$ e $\partial A$.

**Ragionamento:**

**Passo 1 — interno.** Ogni punto del disco aperto $\{x^2+y^2<1\}$ ha una palla attorno interamente dentro il disco (per definizione di disco aperto), quindi dentro $A$: questi punti sono interni. Il punto $(1,0)$ **non** è interno: ogni palla attorno a $(1,0)$, per quanto piccola, contiene punti con $x^2+y^2>1$ che non stanno in $A$ (l'unico punto di $A$ sulla circonferenza è $(1,0)$ stesso). Dunque $\operatorname{int}(A)=\{x^2+y^2<1\}$ (il disco aperto, senza il punto isolato).

**Passo 2 — chiusura.** Il disco aperto ha come punti di accumulazione tutti i punti della circonferenza $x^2+y^2=1$ (ogni intorno di un punto sulla circonferenza contiene punti del disco aperto, avvicinandosi da dentro) — questo è un fatto standard sulla chiusura delle palle. Il punto $(1,0)$ è già in $A$. Nessun punto con $x^2+y^2>1$ è di accumulazione (ha un intorno interamente fuori da $A$, essendo $A$ contenuto nel disco chiuso). Dunque $\overline A=\{x^2+y^2\le1\}$ (il disco chiuso — il punto isolato $(1,0)$ non aggiunge nulla, essendo già sulla circonferenza).

**Passo 3 — frontiera.** $\partial A=\overline A\setminus\operatorname{int}(A)=\{x^2+y^2\le1\}\setminus\{x^2+y^2<1\}=\{x^2+y^2=1\}$: l'intera circonferenza, non solo il punto $(1,0)$.

**Osservazione:** questo esempio mostra che aggiungere un singolo punto isolato sulla frontiera del disco **non cambia né la chiusura né la frontiera** (che dipendono dalla "forma complessiva" vicino a ogni punto), ma toglie quel punto dall'interno. È un buon esercizio per capire che interno, chiusura e frontiera sono nozioni **locali** (dipendono da cosa succede in ogni intorno), non globali.

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 5))
theta = np.linspace(0, 2*np.pi, 300)

# Disco aperto (tratteggiato sul bordo, per indicare "escluso")
ax.fill(np.cos(theta), np.sin(theta), color='lightblue', alpha=0.6, label='int(A) = disco aperto')
ax.plot(np.cos(theta), np.sin(theta), 'r-', linewidth=2, label="frontiera = intera circonferenza")
ax.plot(1, 0, 'go', markersize=10, label="punto isolato (1,0), in A ma non interno")

ax.set_aspect('equal')
ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
ax.set_title("A = disco aperto unitario U {(1,0)}")
plt.tight_layout()
plt.show()

print("int(A): disco aperto x^2+y^2 < 1 (il punto isolato (1,0) NON e' interno)")
print("chiusura(A): disco chiuso x^2+y^2 <= 1")
print("frontiera(A): l'intera circonferenza x^2+y^2 = 1")
```

La seconda condizione dice che nessun aperto contenente $x$ è contenuto in $A$, cioè ogni intorno di $x$ contiene punti fuori da $A$, cioè punti di $X\setminus A$: questo significa $x\in\overline{X\setminus A}$ (essendo $x\in X\setminus A$ oppure punto di accumulazione di $X\setminus A$). Dunque $x\in\overline A\cap\overline{X\setminus A}$. Il ragionamento è reversibile in ogni passo, dando l'uguaglianza degli insiemi.

**3.** *(⇒)* Se $A$ è aperto, ogni $x\in A$ ha un intorno (se stesso, essendo aperto) contenuto in $A$, quindi $x\in\operatorname{int}(A)$, cioè $A\subseteq\operatorname{int}(A)$; sempre $\operatorname{int}(A)\subseteq A$, quindi $A=\operatorname{int}(A)$, e $\partial A=\overline A\setminus A$, che non può intersecare $A$ (per definizione di differenza insiemistica). *(⇐)* Se $A\cap\partial A=\emptyset$: per ogni $x\in A$, essendo $x\in\overline A$ (sempre vero) e $x\notin\partial A$ per ipotesi, deve essere $x\in\operatorname{int}(A)$ (dato che $\partial A=\overline A\setminus\operatorname{int}(A)$, $x\in\overline A\setminus\partial A\Rightarrow x\in\operatorname{int}(A)$). Dunque $A\subseteq\operatorname{int}(A)\subseteq A$, cioè $A=\operatorname{int}(A)$, che è aperto (essendo unione di aperti per definizione).

**4.** Nella topologia discreta, ogni sottoinsieme è aperto (per definizione, $\tau=\mathcal P(X)$). Combinando: $\overline A=A=\operatorname{int}(A)$, quindi $\partial A=\overline A\setminus\operatorname{int}(A)=A\setminus A=\emptyset$.

## Riepilogo finale

- Un insieme **chiuso** è il complementare di un aperto; per De Morgan, i chiusi sono stabili per intersezioni arbitrarie e unioni finite (duale degli aperti).
- La **chiusura** $\overline A$ (il più piccolo chiuso contenente $A$) si ottiene aggiungendo ad $A$ i suoi punti di accumulazione; l'**interno** $\operatorname{int}(A)$ (il più grande aperto dentro $A$) toglie i punti "di confine".
- La **frontiera** $\partial A=\overline A\setminus\operatorname{int}(A)$ raccoglie i punti che stanno "al limite" tra $A$ e il suo complementare.
- $A$ è chiuso $\iff$ $A=\overline A$ $\iff$ $A$ contiene tutti i suoi punti di accumulazione; $A$ è aperto $\iff$ $A\cap\partial A=\emptyset$.
