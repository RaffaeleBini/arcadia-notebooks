# 12.9 — Spazi prodotto

**Prerequisiti:** 12.8 Completezza (spazi metrici), 12.4 Continuità (topologica).


## Perché ci interessa

$\mathbb R^n$ non è altro che il prodotto cartesiano $\mathbb R\times\cdots\times\mathbb R$ ($n$ volte), e nel notebook 12.8 abbiamo dimostrato la sua completezza *riducendola componente per componente* a quella di $\mathbb R$. Questa tecnica — studiare un prodotto tramite le sue componenti — funziona perché $\mathbb R^n$ ha una topologia "compatibile" con la struttura di prodotto: le proiezioni sulle singole coordinate sono continue, e la convergenza in $\mathbb R^n$ equivale alla convergenza di ogni coordinata separatamente. Questo notebook rende esplicita la costruzione generale — la **topologia prodotto** — così che tecniche come quella del notebook 12.8 si applichino automaticamente a *qualunque* prodotto di spazi topologici, non solo a $\mathbb R^n$.

## Teoria, passo dopo passo

### Definizione di topologia prodotto (caso finito)

Dati spazi topologici $(X_1,\tau_1),\ldots,(X_n,\tau_n)$, la **topologia prodotto** su $X_1\times\cdots\times X_n$ è quella generata dalla base $\mathcal B=\{U_1\times\cdots\times U_n:U_i\in\tau_i\}$ (i "rettangoli" con lati aperti in ciascun fattore). Si verifica facilmente (analogo all'esercizio del notebook 12.2) che $\mathcal B$ soddisfa le condizioni per essere base: l'intersezione di due rettangoli è ancora un rettangolo, $(U_1\times\cdots\times U_n)\cap(V_1\times\cdots\times V_n)=(U_1\cap V_1)\times\cdots\times(U_n\cap V_n)$.

**Coerenza con $\mathbb R^n$:** la topologia prodotto su $\mathbb R\times\cdots\times\mathbb R$ ($n$ volte, ciascuno con la topologia euclidea) coincide con la topologia euclidea di $\mathbb R^n$: questo è precisamente il contenuto dell'esercizio svolto del notebook 12.2, dove i rettangoli aperti $(a,b)\times(c,d)$ generavano la stessa topologia delle palle in $\mathbb R^2$ — un caso particolare ($n=2$) di questo fatto generale.

### Proiezioni continue e proprietà universale

Le **proiezioni** $\pi_i:X_1\times\cdots\times X_n\to X_i$, $\pi_i(x_1,\ldots,x_n)=x_i$, sono sempre continue nella topologia prodotto.

**Dimostrazione.** Sia $U_i\in\tau_i$. $\pi_i^{-1}(U_i)=X_1\times\cdots\times X_{i-1}\times U_i\times X_{i+1}\times\cdots\times X_n$ — un elemento della base (con tutti i fattori uguali all'intero spazio tranne l'$i$-esimo), quindi aperto. $\blacksquare$ Anzi, la topologia prodotto è **la più povera** (con meno aperti) tra tutte quelle che rendono continue tutte le proiezioni — è la costruzione "minimale" con questa proprietà, un fatto che in topologia più avanzata si chiama proprietà universale del prodotto.

### Continuità verso un prodotto: basta controllare le componenti

**Teorema.** $f:Y\to X_1\times\cdots\times X_n$ è continua se e solo se ogni composizione $\pi_i\circ f:Y\to X_i$ è continua.

**Dimostrazione.** *(⇒)* Se $f$ è continua, $\pi_i\circ f$ è composizione di continue (notebook 12.4), quindi continua. *(⇐)* Supponiamo ogni $\pi_i\circ f$ continua. Sia $U_1\times\cdots\times U_n$ un elemento di base del prodotto. $f^{-1}(U_1\times\cdots\times U_n)=\{y:f(y)\in U_1\times\cdots\times U_n\}=\{y:\pi_i(f(y))\in U_i\;\forall i\}=\bigcap_{i=1}^n(\pi_i\circ f)^{-1}(U_i)$ — intersezione **finita** di aperti (ciascuno aperto per continuità di $\pi_i\circ f$), quindi aperta. Essendo la controimmagine di ogni elemento di base aperta, e ogni aperto del prodotto unione di elementi di base, $f^{-1}$ di ogni aperto è aperta (unione di aperti): $f$ è continua. $\blacksquare$

**Corollario pratico:** una curva $\gamma:\mathbb R\to\mathbb R^n$, $\gamma(t)=(\gamma_1(t),\ldots,\gamma_n(t))$, è continua se e solo se ogni componente $\gamma_i(t)$ è continua — il fatto, spesso usato senza dimostrazione nei corsi di analisi, che per verificare la continuità di una curva basta controllarla coordinata per coordinata.

### Compattezza del prodotto: teorema di Tychonoff (caso finito)

**Teorema.** Se $X_1,\ldots,X_n$ sono compatti, allora $X_1\times\cdots\times X_n$ (con la topologia prodotto) è compatto.

Questo è un caso particolare del **teorema di Tychonoff**, che vale anche per prodotti infiniti (un risultato profondo, equivalente all'assioma della scelta nella sua forma generale) — qui ci limitiamo al caso finito, dimostrabile per induzione a partire dal caso $n=2$ usando il **lemma del tubo** (tecnica che esula da questo notebook introduttivo, ma il risultato per $n=2$ è essenzialmente equivalente al teorema di Heine-Borel su $\mathbb R^2$, già visto nel notebook 10.11 e 12.6, generalizzato).

```python
import numpy as np
import matplotlib.pyplot as plt

# Illustriamo la topologia prodotto: un elemento di base U1 x U2 in R^2, e le proiezioni
fig, ax = plt.subplots(figsize=(6, 6))

U1 = (1, 3)  # intervallo aperto su asse x
U2 = (0.5, 2.5)  # intervallo aperto su asse y

rect_x = [U1[0], U1[1], U1[1], U1[0], U1[0]]
rect_y = [U2[0], U2[0], U2[1], U2[1], U2[0]]
ax.fill(rect_x, rect_y, alpha=0.3, color='steelblue', label='U1 x U2 (elemento di base)')
ax.plot(rect_x, rect_y, 'b-')

ax.axvspan(U1[0], U1[1], ymin=0, ymax=0.05, color='red', label='proiezione su asse x = U1')
ax.axhspan(U2[0], U2[1], xmin=0, xmax=0.05, color='green', label='proiezione su asse y = U2')

ax.set_xlim(0, 4)
ax.set_ylim(0, 3.5)
ax.legend(loc='upper right', fontsize=8)
ax.set_title("Topologia prodotto su R^2: base = rettangoli U1 x U2")
plt.tight_layout()
plt.show()

# Verifica: la proiezione di un rettangolo aperto sulla prima coordinata e' un intervallo aperto
punti_nel_rettangolo = [(x, y) for x in np.linspace(*U1, 5) for y in np.linspace(*U2, 5)]
proiezioni_x = sorted(set(round(p[0], 2) for p in punti_nel_rettangolo))
print("Campionamento delle proiezioni sulla prima coordinata:", proiezioni_x)
print(f"Range atteso: {U1}")
```

## Esercizio svolto e commentato

**Problema:** dimostra che, nella topologia prodotto su $X\times Y$ (con $X,Y$ spazi metrici), la convergenza di una successione $(x_n,y_n)\to(x,y)$ equivale alla convergenza componente per componente: $x_n\to x$ **e** $y_n\to y$.

**Ragionamento:**

**Passo 1 — usiamo la metrica prodotto.** Dotiamo $X\times Y$ della metrica $d((x_1,y_1),(x_2,y_2))=\max(d_X(x_1,x_2),d_Y(y_1,y_2))$ (analoga alla metrica $d_\infty$ dell'esercizio 3 del notebook 12.7): si verifica che questa metrica induce esattamente la topologia prodotto (le palle di questa metrica sono $B_X(x,r)\times B_Y(y,r)$, elementi di base del prodotto).

**Passo 2 — dimostriamo l'implicazione (⇒).** Se $(x_n,y_n)\to(x,y)$ nella metrica prodotto, per ogni $\varepsilon>0$ esiste $N$ con $d((x_n,y_n),(x,y))<\varepsilon$ per $n\ge N$, cioè $\max(d_X(x_n,x),d_Y(y_n,y))<\varepsilon$. Questo implica **sia** $d_X(x_n,x)<\varepsilon$ **sia** $d_Y(y_n,y)<\varepsilon$ (essendo entrambi $\le$ del massimo): dunque $x_n\to x$ e $y_n\to y$ separatamente.

**Passo 3 — dimostriamo l'implicazione (⇐).** Se $x_n\to x$ e $y_n\to y$: dato $\varepsilon>0$, esiste $N_1$ con $d_X(x_n,x)<\varepsilon$ per $n\ge N_1$, ed $N_2$ con $d_Y(y_n,y)<\varepsilon$ per $n\ge N_2$. Prendendo $N=\max(N_1,N_2)$, per $n\ge N$ entrambe le condizioni valgono, quindi $\max(d_X(x_n,x),d_Y(y_n,y))<\varepsilon$, cioè $d((x_n,y_n),(x,y))<\varepsilon$: $(x_n,y_n)\to(x,y)$.

**Conclusione:** le due nozioni di convergenza coincidono. **Collegamento:** questo esercizio è la versione "successioni" del teorema della teoria sulla continuità verso un prodotto (che ne è, in un certo senso, la generalizzazione: una successione è una funzione $\mathbb N\to X\times Y$, e la sua "continuità nel punto all'infinito" è proprio la convergenza).

```python
import numpy as np

# Verifica numerica: convergenza in R^2 (prodotto R x R) equivale a convergenza per componenti
n_vals = np.arange(1, 100)
x_n = 1 + 1/n_vals          # converge a 1
y_n = 2 - 1/n_vals**2       # converge a 2

punti = np.stack([x_n, y_n], axis=1)
limite = np.array([1, 2])

distanze_prodotto = np.max(np.abs(punti - limite), axis=1)  # metrica del massimo
distanze_x = np.abs(x_n - limite[0])
distanze_y = np.abs(y_n - limite[1])

print("n=10:", "d_prodotto =", distanze_prodotto[9], "d_x =", distanze_x[9], "d_y =", distanze_y[9])
print("n=99:", "d_prodotto =", distanze_prodotto[98], "d_x =", distanze_x[98], "d_y =", distanze_y[98])
print("\nTutte e tre le distanze tendono a 0 insieme: convergenza nel prodotto <=> convergenza per componenti")
```

## Riepilogo finale

- La **topologia prodotto** su $X_1\times\cdots\times X_n$ è generata dai "rettangoli" $U_1\times\cdots\times U_n$: coincide, per $n$ copie di $\mathbb R$, con la topologia euclidea di $\mathbb R^n$.
- Le proiezioni sono sempre continue, ed è proprio questa proprietà a caratterizzare la topologia prodotto come la "più povera" che le rende tali.
- Una funzione verso un prodotto è continua se e solo se lo sono tutte le sue componenti: questo giustifica il controllo "componente per componente" usato di continuo in analisi.
- Compattezza, connessione e chiusura si comportano bene rispetto ai prodotti finiti: prodotti di compatti sono compatti, prodotti di connessi sono connessi, prodotti di chiusi sono chiusi.
