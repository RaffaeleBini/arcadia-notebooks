## Perché ci interessa

"Sconto del 20%", "tasso d'interesse del 3%", "il 45% degli elettori": le percentuali sono ovunque, perché offrono un modo comune per confrontare grandezze diverse tra loro, indipendentemente dalla scala dei numeri in gioco. Dire "ho risparmiato 20 euro" dice poco se non sappiamo su quale prezzo di partenza; dire "ho risparmiato il 20%" è immediatamente confrontabile con qualsiasi altro sconto.

Le percentuali non sono un argomento nuovo: sono semplicemente un caso particolare dei rapporti visti nel notebook precedente, con una scelta di denominatore molto comoda: il 100.

## Teoria, passo dopo passo

### Cos'è una percentuale

"Per cento" (percentuale, dal latino *per centum*) significa "ogni cento". Il simbolo $\%$ è solo un modo compatto di scrivere "diviso 100":

$$p\% = \frac{p}{100}$$

Per esempio $25\% = \frac{25}{100} = \frac{1}{4} = 0.25$. Dire "il 25% di una quantità" equivale a moltiplicare quella quantità per $0.25$.

### Calcolare la percentuale di una quantità

Per calcolare il $p\%$ di una quantità $Q$:

$$\text{parte} = Q \times \frac{p}{100}$$

Esempio: il 20% di 150 è $150 \times \frac{20}{100} = 150 \times 0.2 = 30$.

Questo è in realtà un caso particolare della proporzione vista nel notebook 0.9: $\text{parte} : Q = p : 100$.

### Le tre domande tipiche

Con le percentuali si presentano quasi sempre tre tipi di domande, tutte risolvibili con la stessa proporzione $\text{parte}:Q = p:100$, cambiando solo cosa è incognito:

1. **Trovare la parte, note $Q$ e $p$:** "quanto è il 20% di 150?" → $\text{parte} = \frac{150 \times 20}{100}=30$.
2. **Trovare la percentuale, note $Q$ e la parte:** "30 è che percentuale di 150?" → $p = \frac{30 \times 100}{150} = 20\%$.
3. **Trovare il totale, note la parte e $p$:** "30 è il 20% di quale numero?" → $Q = \frac{30\times100}{20}=150$.

Non servono tre formule diverse da imparare a memoria: basta ricordare l'unica proporzione $\text{parte}:Q=p:100$ e usare la proprietà fondamentale delle proporzioni (notebook 0.9) per isolare l'incognita di volta in volta.

### Aumenti e diminuzioni percentuali

Se una quantità $Q$ **aumenta** del $p\%$, il nuovo valore è:

$$Q_{\text{nuovo}} = Q + Q\times\frac{p}{100} = Q\left(1+\frac{p}{100}\right)$$

Se **diminuisce** del $p\%$:

$$Q_{\text{nuovo}} = Q - Q\times\frac{p}{100} = Q\left(1-\frac{p}{100}\right)$$

**Attenzione a un errore molto comune**: se un prezzo aumenta del 10% e poi diminuisce del 10%, **non** torna al valore originale! Aumentare del 10% significa moltiplicare per $1.1$; diminuire del 10% significa moltiplicare per $0.9$. Il fattore combinato è $1.1 \times 0.9 = 0.99$, cioè una diminuzione netta dell'1% rispetto al valore di partenza. Questo succede perché il secondo 10% si calcola su una base diversa (più grande) rispetto al primo.

````python
import sympy as sp

# Le tre domande tipiche sulle percentuali
Q = 150
p = sp.Rational(20)  # 20%

# 1. Trovare la parte
parte = Q * p / 100
print(f"Il {p}% di {Q} è {parte}")

# 2. Trovare la percentuale
parte2 = 30
percentuale = parte2 * 100 / Q
print(f"{parte2} è il {percentuale}% di {Q}")

# 3. Trovare il totale
parte3 = 30
p3 = sp.Rational(20)
totale = parte3 * 100 / p3
print(f"{parte3} è il {p3}% di {totale}")

# Verifica dell'insidia aumento+diminuzione
prezzo_iniziale = 100
dopo_aumento = prezzo_iniziale * sp.Rational(110, 100)   # +10%
dopo_diminuzione = dopo_aumento * sp.Rational(90, 100)   # -10%
print(f"\nPrezzo iniziale: {prezzo_iniziale}")
print(f"Dopo +10%: {dopo_aumento}")
print(f"Dopo -10% (sul nuovo valore): {dopo_diminuzione}")
print(f"Variazione netta rispetto all'inizio: {(dopo_diminuzione - prezzo_iniziale) / prezzo_iniziale * 100}%")
````

## Esercizio svolto e commentato

**Problema:** un articolo costa 80 euro. Viene scontato del 15%, e sul prezzo scontato viene applicata un'IVA del 22%. Qual è il prezzo finale?

**Ragionamento:**

Procediamo per passi, applicando ogni variazione percentuale **sul valore ottenuto al passo precedente** (non sul prezzo originale, un errore comune).

**Passo 1 — sconto del 15% su 80 euro:**

Uno sconto del 15% significa moltiplicare per $\left(1-\frac{15}{100}\right)=0.85$:

$$80 \times 0.85 = 68 \text{ euro}$$

**Passo 2 — IVA del 22% sui 68 euro** (non sugli 80 originali, perché l'IVA si applica al prezzo già scontato):

Un aumento del 22% significa moltiplicare per $\left(1+\frac{22}{100}\right)=1.22$:

$$68 \times 1.22 = 82.96 \text{ euro}$$

Il prezzo finale è **82.96 euro**.

**Perché non possiamo semplicemente fare "sconto 15% - IVA 22% = +7%" e applicarlo direttamente:** le percentuali di aumento e diminuzione non si sommano algebricamente quando si applicano in sequenza, perché ogni percentuale è calcolata su una base diversa (prima 80, poi 68). Il fattore combinato corretto è $0.85 \times 1.22 = 1.037$, cioè un aumento netto del $3.7\%$ rispetto al prezzo originale — non del $7\%$ come si potrebbe pensare sommando ingenuamente $-15+22$.

````python
import sympy as sp

prezzo_iniziale = 80
dopo_sconto = prezzo_iniziale * sp.Rational(85, 100)    # -15%
prezzo_finale = dopo_sconto * sp.Rational(122, 100)      # +22%
print(f"Prezzo dopo lo sconto: {float(dopo_sconto)} euro")
print(f"Prezzo finale con IVA: {float(prezzo_finale)} euro")

variazione_netta = (prezzo_finale - prezzo_iniziale) / prezzo_iniziale * 100
print(f"Variazione netta rispetto al prezzo iniziale: {float(variazione_netta):.2f}%")
````

**1.** $240 \times \frac{35}{100} = 84$.

**2.** $\frac{15}{25}\times100 = 60\%$.

**3.** $50 \times 1.2 = 60$ (dopo l'aumento). $60 \times 0.8 = 48$ (dopo lo sconto). Il prezzo finale è **48 euro**, non 50: è diminuito, per lo stesso motivo visto nella teoria (le percentuali si applicano su basi diverse).

**4.** Il fattore combinato è $\left(1+\frac{p}{100}\right)\left(1-\frac{p}{100}\right)$. Questo è un prodotto della forma $(1+x)(1-x)$ con $x=\frac{p}{100}$, che sviluppato dà $1-x^2$ (prodotto notevole, lo vedremo formalmente nel notebook 1.2). Quindi il fattore combinato è $1-\left(\frac{p}{100}\right)^2$. Poiché $\left(\frac{p}{100}\right)^2 \ge 0$ sempre (un quadrato non è mai negativo), il fattore è sempre $\le 1$, con uguaglianza solo quando $p=0$. Quindi il valore finale è sempre $\le$ al valore di partenza, tranne quando $p=0$ (nessuna variazione).

## Riepilogo finale

- $p\% = \frac{p}{100}$: una percentuale è solo un rapporto con denominatore fissato a 100.
- Tutte le domande sulle percentuali si riconducono all'unica proporzione $\text{parte}:Q=p:100$.
- Un aumento del $p\%$ corrisponde a moltiplicare per $\left(1+\frac{p}{100}\right)$; una diminuzione, a moltiplicare per $\left(1-\frac{p}{100}\right)$.
- Variazioni percentuali applicate in sequenza **non si sommano**: vanno moltiplicate, perché ogni percentuale è calcolata su una base diversa.
- Un aumento seguito da una diminuzione della stessa percentuale non riporta mai al valore originale (tranne che con $p=0$): il risultato è sempre minore.
