## Perché ci interessa

Nel notebook precedente abbiamo visto che $\mathbb{Q}$ è denso: tra due razionali distinti ce n'è sempre un altro. Verrebbe da pensare che i razionali "riempiano" completamente la retta numerica. Sorprendentemente, non è così: esistono lunghezze molto concrete, come la diagonale di un quadrato di lato 1, che **non corrispondono a nessun numero razionale**.

Ci serve un insieme di numeri più grande, che includa anche questi "buchi": l'insieme dei **numeri reali**.

## Teoria, passo dopo passo

### Un esempio concreto: la diagonale del quadrato

Prendi un quadrato di lato 1. Per il teorema di Pitagora, la sua diagonale misura $\sqrt{1^2+1^2} = \sqrt{2}$. È una lunghezza reale, misurabile con un righello. Ma $\sqrt{2}$ è un numero razionale?

**Dimostriamo che $\sqrt{2}$ non è razionale**, con un ragionamento per assurdo (lo vedremo in dettaglio nel notebook 0.20, qui ne diamo un'anteprima concreta):

Supponiamo, per assurdo, che $\sqrt{2} = \frac{a}{b}$ con $a, b$ interi senza fattori comuni (frazione ridotta ai minimi termini). Elevando al quadrato:

$$2 = \frac{a^2}{b^2} \quad \Rightarrow \quad a^2 = 2b^2$$

Quindi $a^2$ è pari (è il doppio di $b^2$). Ma se $a^2$ è pari, allora anche $a$ deve essere pari (il quadrato di un dispari è sempre dispari, come si può verificare: $1^2=1$, $3^2=9$, $5^2=25$...). Scriviamo $a = 2k$. Sostituendo:

$$(2k)^2 = 2b^2 \quad \Rightarrow \quad 4k^2 = 2b^2 \quad \Rightarrow \quad b^2 = 2k^2$$

Quindi anche $b^2$ è pari, e quindi anche $b$ è pari. Ma allora sia $a$ che $b$ sono pari, cioè hanno il fattore comune 2 — **in contraddizione** con l'ipotesi di partenza che $a$ e $b$ non avessero fattori comuni!

Questa contraddizione ci dice che l'ipotesi iniziale era sbagliata: $\sqrt{2}$ **non può essere scritto** come frazione di interi. Non è un limite della nostra bravura nel cercare la frazione giusta: è impossibile in linea di principio.

### Numeri irrazionali e numeri reali

Un numero come $\sqrt{2}$, che non si può scrivere come rapporto di interi, si chiama **numero irrazionale**. Altri esempi famosi: $\pi$, $e$ (la base dei logaritmi naturali, che vedremo nel Modulo 1), $\sqrt{3}$, $\sqrt{5}$.

L'insieme dei **numeri reali** $\mathbb{R}$ è l'unione dei razionali e degli irrazionali: è l'insieme di **tutti** i punti della retta numerica, senza buchi.

$$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$$

### Rappresentazione decimale

Un modo pratico per distinguere razionali e irrazionali è guardare la loro rappresentazione decimale:

- I numeri razionali hanno una rappresentazione decimale **finita** (es. $\frac{1}{4}=0.25$) oppure **periodica** (es. $\frac{1}{3}=0.3333\dots = 0.\overline{3}$).
- I numeri irrazionali hanno una rappresentazione decimale **infinita e non periodica**: le cifre non si ripetono mai secondo uno schema fisso. Per esempio $\sqrt{2} = 1.41421356\dots$

### Perché "riempire i buchi" è importante

L'idea intuitiva di $\mathbb{R}$ come "retta senza buchi" si formalizza con la proprietà di **completezza**, che studieremo con rigore nel Modulo 5 (Analisi Matematica I). Per ora ci basta l'immagine intuitiva: ogni punto della retta corrisponde a un numero reale, e viceversa. Questa proprietà è cruciale: è ciò che renderà possibile definire concetti come limite e continuità più avanti nel corso.

````python
import math
from fractions import Fraction

# Verifichiamo numericamente che sqrt(2) non sembra "stabilizzarsi" su nessuna frazione semplice
radice_2 = math.sqrt(2)
print(f"sqrt(2) ≈ {radice_2}")

# Proviamo ad approssimare sqrt(2) con frazioni sempre più precise
approssimazioni = [Fraction(14, 10), Fraction(141, 100), Fraction(1414, 1000), Fraction(14142, 10000)]
for frazione in approssimazioni:
    valore = float(frazione)
    errore = abs(valore - radice_2)
    print(f"{frazione} = {valore:.6f}   errore rispetto a sqrt(2): {errore:.6f}")
print("\nOgni approssimazione è razionale, ma nessuna è ESATTAMENTE sqrt(2): l'errore si riduce ma non si annulla mai.")

# Confrontiamo le rappresentazioni decimali: finita, periodica, infinita non periodica
print(f"\n1/4 = {Fraction(1,4)} -> decimale finito: {1/4}")
print(f"1/3 = {Fraction(1,3)} -> decimale periodico: {1/3:.10f}...")
print(f"sqrt(2) -> decimale infinito non periodico: {radice_2:.10f}...")
````

## Esercizio svolto e commentato

**Problema:** dimostra, con lo stesso schema di ragionamento usato per $\sqrt{2}$, che $\sqrt{3}$ non è razionale.

**Ragionamento:**

Riprendiamo esattamente la stessa strategia: dimostrazione per assurdo.

Supponiamo che $\sqrt{3} = \frac{a}{b}$, con $a,b$ interi senza fattori comuni. Elevando al quadrato:

$$3 = \frac{a^2}{b^2} \quad \Rightarrow \quad a^2 = 3b^2$$

Quindi $a^2$ è multiplo di 3. Questo implica che anche $a$ è multiplo di 3 (fatto analogo a "il quadrato di un pari è pari": qui usiamo che se $a$ non fosse multiplo di 3, nemmeno $a^2$ lo sarebbe — puoi verificarlo provando $a=3k+1$ o $a=3k+2$ e vedendo che $a^2$ non è mai multiplo di 3 in questi casi).

Scriviamo allora $a = 3k$. Sostituendo:

$$(3k)^2 = 3b^2 \quad \Rightarrow \quad 9k^2 = 3b^2 \quad \Rightarrow \quad b^2 = 3k^2$$

Quindi anche $b^2$, e di conseguenza $b$, sono multipli di 3. Ma allora $a$ e $b$ hanno entrambi il fattore comune 3, **contraddicendo** l'ipotesi che non ne avessero.

Quindi $\sqrt{3}$ non è razionale.

**Perché lo stesso schema funziona:** l'argomento usato per $\sqrt{2}$ non dipendeva davvero dal numero 2, ma dal fatto che 2 (e allo stesso modo 3) sono numeri primi. Questo è un esempio di come, in matematica, capire *la struttura* di una dimostrazione (non solo il caso specifico) permette di riusarla per problemi simili.

````python
import math

def è_probabilmente_irrazionale(n, precisione=10**6):
    # Verifica euristica (non una dimostrazione!): controlla se n è un quadrato perfetto
    radice = math.isqrt(n)
    return radice * radice != n

for n in [2, 3, 4, 5, 9]:
    print(f"sqrt({n}) è irrazionale? {è_probabilmente_irrazionale(n)}")
# Nota: questo codice NON dimostra nulla, verifica solo se n è un quadrato perfetto.
# La dimostrazione rigorosa è quella scritta sopra, a parole.
````

## Riepilogo finale

- Esistono lunghezze reali, come $\sqrt{2}$, che non si possono scrivere come frazione di interi: sono i **numeri irrazionali**.
- L'insieme dei **numeri reali** $\mathbb{R}$ unisce razionali e irrazionali: rappresenta tutti i punti della retta numerica.
- $\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$: ogni insieme numerico visto finora è contenuto nel successivo.
- I razionali hanno rappresentazione decimale finita o periodica; gli irrazionali hanno rappresentazione decimale infinita e non periodica.
- La dimostrazione per assurdo (usata per $\sqrt{2}$ e $\sqrt{3}$) è una tecnica potente: si suppone il contrario di ciò che si vuole dimostrare e si mostra che porta a una contraddizione.
