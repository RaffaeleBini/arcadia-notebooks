## Perché ci interessa

Finora abbiamo parlato di numeri (naturali, interi, razionali, reali) come se fossero "collezioni" di elementi: $\mathbb{N}=\{0,1,2,3,\dots\}$. In realtà abbiamo già usato, in modo informale, il concetto di **insieme**: una collezione di oggetti ben definita.

Ora rendiamo questo concetto esplicito e rigoroso. Gli insiemi sono il "linguaggio" più basilare della matematica moderna: quasi ogni oggetto matematico (numeri, funzioni, punti geometrici...) può essere descritto in termini di insiemi. Capirli bene ora renderà tutto il resto del corso più chiaro.

## Teoria, passo dopo passo

### Cos'è un insieme

Un **insieme** è una collezione di oggetti distinti, chiamati **elementi**, tale che per ogni oggetto si possa dire con certezza se appartiene o no all'insieme. Per esempio: "l'insieme dei giorni della settimana" è un insieme ben definito (7 elementi); "l'insieme delle persone alte" non è un insieme ben definito nel senso matematico, perché "alte" non ha un criterio preciso (alto quanto?).

Indichiamo gli insiemi con lettere maiuscole ($A$, $B$, $\mathbb{N}$...) e gli elementi con lettere minuscole. Se $x$ è un elemento di $A$, scriviamo:

$$x \in A \qquad \text{(si legge "$x$ appartiene ad $A$")}$$

Se $x$ non è un elemento di $A$:

$$x \notin A$$

### Due modi per descrivere un insieme

**Per elencazione** (si elencano tutti gli elementi, tra parentesi graffe):

$$A = \{1, 2, 3, 4, 5\}$$

**Per proprietà caratteristica** (si descrive una condizione che gli elementi devono soddisfare):

$$A = \{n \in \mathbb{N} : 1 \le n \le 5\}$$

si legge "l'insieme degli $n$ appartenenti a $\mathbb{N}$ tali che $1\le n\le 5$". I due punti (a volte si usa una barra verticale $|$) si leggono "tali che". Questa seconda notazione è indispensabile quando l'insieme ha infiniti elementi, ed è la notazione che useremo sempre di più andando avanti nel corso.

### Insiemi speciali

- **Insieme vuoto**, indicato $\emptyset$ (o $\{\}$): l'insieme che non ha nessun elemento. Esempio: $\{n\in\mathbb{N} : n<0\} = \emptyset$ (nessun naturale è negativo).
- Un insieme è **finito** se ha un numero finito di elementi (es. $\{1,2,3\}$), altrimenti è **infinito** (es. $\mathbb{N}$).

### Sottoinsiemi

$A$ è un **sottoinsieme** di $B$, scritto $A \subseteq B$, se ogni elemento di $A$ è anche elemento di $B$:

$$A \subseteq B \quad \Longleftrightarrow \quad \text{per ogni } x,\ (x \in A \Rightarrow x \in B)$$

Esempio: $\{1,2\} \subseteq \{1,2,3\}$, perché sia $1$ che $2$ appartengono anche al secondo insieme. Ogni insieme è sottoinsieme di sé stesso ($A\subseteq A$), e l'insieme vuoto è sottoinsieme di ogni insieme ($\emptyset \subseteq A$, per ogni $A$) — questo può sembrare strano, ma è vero "per vacuità": non esiste nessun elemento di $\emptyset$ che possa violare la condizione, quindi la condizione è automaticamente soddisfatta.

Se $A\subseteq B$ ma $A \ne B$ (esiste almeno un elemento di $B$ che non è in $A$), diciamo che $A$ è un **sottoinsieme proprio** di $B$, e scriviamo $A \subsetneq B$.

### Uguaglianza tra insiemi

Due insiemi sono **uguali** se hanno esattamente gli stessi elementi (l'ordine e le ripetizioni non contano: $\{1,2,3\}=\{3,1,2\}=\{1,1,2,3\}$). Un modo rigoroso per dimostrare $A=B$ è mostrare **due inclusioni**:

$$A = B \quad \Longleftrightarrow \quad A\subseteq B \ \text{ e } \ B\subseteq A$$

Questo metodo di dimostrazione ("doppia inclusione") tornerà molto utile più avanti nel corso.

````python
# Python ha un tipo nativo per gli insiemi: set
A = {1, 2, 3, 4, 5}
B = {1, 2, 3}

print("A =", A)
print("3 in A?", 3 in A)   # corrisponde a "3 ∈ A"
print("9 in A?", 9 in A)   # corrisponde a "9 ∉ A"

# Sottoinsieme
print("\nB è sottoinsieme di A?", B.issubset(A))   # corrisponde a "B ⊆ A"
print("A è sottoinsieme di B?", A.issubset(B))

# Insieme vuoto è sottoinsieme di qualunque insieme
vuoto = set()
print("\ninsieme vuoto sottoinsieme di A?", vuoto.issubset(A))

# Insieme per proprietà caratteristica (comprensione in Python, molto simile alla notazione matematica)
C = {n for n in range(0, 20) if n % 3 == 0}  # {n in N : n multiplo di 3, n<20}
print("\nC = {n in N, n<20 : n multiplo di 3} =", C)

# Uguaglianza tra insiemi: l'ordine e le ripetizioni non contano
print("\n{1,2,3} == {3,1,2}?", {1, 2, 3} == {3, 1, 2})
````

## Esercizio svolto e commentato

**Problema:** siano $A=\{n \in \mathbb{N} : n \text{ è pari e } n \le 10\}$ e $B=\{0,2,4,6,8,10\}$. Dimostra che $A=B$.

**Ragionamento:**

Usiamo il metodo della doppia inclusione visto nella teoria: dobbiamo mostrare sia $A\subseteq B$ sia $B\subseteq A$.

**Dimostriamo $A \subseteq B$:** prendiamo un elemento generico $x \in A$. Per definizione di $A$, $x$ è un naturale pari con $x \le 10$. I naturali pari da $0$ a $10$ sono esattamente $0,2,4,6,8,10$ (li possiamo elencare perché sono un numero finito). Quindi $x$ deve essere uno di questi valori, cioè $x \in B$. Poiché questo vale per un $x$ generico, ogni elemento di $A$ è anche in $B$: $A \subseteq B$.

**Dimostriamo $B \subseteq A$:** prendiamo un elemento generico $y \in B$, cioè $y$ è uno tra $0,2,4,6,8,10$. Ognuno di questi valori è un naturale pari e minore o uguale a 10 (verifica diretta su ciascuno). Quindi $y$ soddisfa la proprietà caratteristica di $A$, cioè $y \in A$. Poiché questo vale per un $y$ generico, $B \subseteq A$.

Avendo mostrato entrambe le inclusioni, concludiamo $A=B$.

**Perché non basta "elencare e confrontare visivamente":** con insiemi piccoli come questo, elencare e confrontare sembra sufficiente. Ma il metodo della doppia inclusione è quello che funziona **sempre**, anche quando gli insiemi sono infiniti o troppo grandi da elencare — è uno strumento che ci accompagnerà per tutto il corso, quindi vale la pena abituarsi a usarlo fin da ora, anche su esempi semplici.

````python
A = {n for n in range(0, 11) if n % 2 == 0}  # n <= 10 pari
B = {0, 2, 4, 6, 8, 10}

print("A =", A)
print("B =", B)
print("A == B?", A == B)
print("A sottoinsieme di B?", A.issubset(B))
print("B sottoinsieme di A?", B.issubset(A))
````

## Riepilogo finale

- Un insieme è una collezione di elementi ben definita; $x\in A$ significa "$x$ appartiene ad $A$".
- Si descrive per elencazione ($\{1,2,3\}$) o per proprietà caratteristica ($\{x : \dots\}$).
- $A\subseteq B$ significa che ogni elemento di $A$ è anche elemento di $B$; l'insieme vuoto è sottoinsieme di qualunque insieme.
- Due insiemi sono uguali se e solo se hanno esattamente gli stessi elementi; il modo rigoroso per dimostrarlo è la **doppia inclusione** ($A\subseteq B$ e $B\subseteq A$).
- Questi concetti (appartenenza, inclusione, uguaglianza) sono il vocabolario di base che useremo in tutto il resto del corso.
