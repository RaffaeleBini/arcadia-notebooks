## Teoria, passo dopo passo

### L'implicazione

L'implicazione $p \Rightarrow q$ (si legge "$p$ implica $q$", oppure "se $p$ allora $q$") ha questa tavola di verità:

| $p$ | $q$ | $p \Rightarrow q$ |
|---|---|---|
| V | V | V |
| V | F | **F** |
| F | V | V |
| F | F | V |

L'unico caso in cui $p\Rightarrow q$ è **falsa** è quando $p$ è vera ma $q$ è falsa. Negli altri tre casi è vera — e questo è il punto che sorprende di più chi vede l'implicazione per la prima volta: se $p$ è **falsa**, l'implicazione è vera **a prescindere** da $q$ ("vera per vacuità"). Esempio concreto: "se la luna è fatta di formaggio, allora $2+2=5$" è considerata vera in logica, perché la premessa è falsa — non ci interessa cosa dice la conclusione quando la premessa non si verifica mai.

In un'implicazione, $p$ si chiama **ipotesi** (o antecedente) e $q$ **tesi** (o conseguente). Questo lessico — ipotesi, tesi — è esattamente quello che useremo scrivendo dimostrazioni a partire dal notebook 0.20.

### Implicazione e sottoinsiemi: un collegamento importante

C'è un legame diretto tra implicazione e inclusione tra insiemi (notebook 0.12): se chiamiamo $A$ l'insieme degli elementi per cui $p$ è vera, e $B$ quello per cui $q$ è vera, allora:

$$p \Rightarrow q \quad \text{(sempre vera)} \qquad \Longleftrightarrow \qquad A \subseteq B$$

Per esempio: "$n$ è multiplo di 4 $\Rightarrow$ $n$ è pari" corrisponde a dire che l'insieme dei multipli di 4 è un sottoinsieme dei numeri pari — che è vero, ed è esattamente il motivo per cui l'implicazione è vera per ogni $n$.

### Come si nega un'implicazione

Un errore comune: pensare che $\neg(p\Rightarrow q)$ sia $\neg p \Rightarrow \neg q$ (o simili). Vediamo la regola corretta, che si può ricavare dalla tavola di verità: $p\Rightarrow q$ è falsa **solo** quando $p$ è vera e $q$ è falsa. Quindi:

$$\neg(p \Rightarrow q) \equiv p \wedge (\neg q)$$

"L'unico modo per smentire 'se $p$ allora $q$' è mostrare un caso in cui $p$ è vera ma $q$ non lo è." Questa regola sarà essenziale quando, nel notebook 0.20, parleremo di controesempi: per confutare un'affermazione "se... allora...", basta trovare **un solo caso** in cui l'ipotesi è vera ma la tesi è falsa.

### L'implicazione inversa e la contronominale

Data $p\Rightarrow q$, si possono costruire altre due proposizioni collegate:

- **Inversa**: $q \Rightarrow p$. **Attenzione**: l'inversa **non** ha lo stesso valore di verità dell'implicazione originale in generale! "$n$ multiplo di 4 $\Rightarrow$ $n$ pari" è vera, ma la sua inversa "$n$ pari $\Rightarrow$ $n$ multiplo di 4" è falsa (per esempio $n=6$ è pari ma non multiplo di 4).
- **Contronominale**: $(\neg q) \Rightarrow (\neg p)$. Questa **è sempre logicamente equivalente** a $p\Rightarrow q$ (puoi verificarlo con la tavola di verità nella cella di codice sotto). Per il nostro esempio: "$n$ non pari $\Rightarrow$ $n$ non multiplo di 4" — vera, esattamente come l'originale. La dimostrazione per contronominale (dimostrare $(\neg q)\Rightarrow(\neg p)$ invece di $p\Rightarrow q$ direttamente) è una tecnica dimostrativa che useremo nel notebook 0.20.

### L'equivalenza (doppia implicazione)

$p \Leftrightarrow q$ (si legge "$p$ se e solo se $q$") significa che $p$ e $q$ hanno **sempre lo stesso valore di verità**: è vera quando entrambe sono vere o entrambe sono false.

$$p \Leftrightarrow q \quad \equiv \quad (p\Rightarrow q) \wedge (q\Rightarrow p)$$

Dimostrare un "se e solo se" richiede quindi due dimostrazioni: una nella direzione $\Rightarrow$, una nella direzione $\Leftarrow$ — esattamente lo stesso schema della doppia inclusione tra insiemi (notebook 0.12).

````python
import itertools

def implica(p, q):
    return (not p) or q  # equivalente a "non p oppure q", stessa tavola di verità di p => q

print(f"{'p':^5}{'q':^5}{'p=>q':^7}{'not(p=>q)':^12}{'p and not q':^12}")
for p, q in itertools.product([True, False], repeat=2):
    non_implica = not implica(p, q)
    p_and_not_q = p and (not q)
    print(f"{str(p):^5}{str(q):^5}{str(implica(p,q)):^7}{str(non_implica):^12}{str(p_and_not_q):^12}")

# Verifica: contronominale sempre equivalente all'originale, inversa no
print("\nConfronto tra originale, inversa e contronominale:")
print(f"{'p':^5}{'q':^5}{'p=>q':^7}{'q=>p (inversa)':^16}{'(not q)=>(not p) (contronom.)':^30}")
for p, q in itertools.product([True, False], repeat=2):
    originale = implica(p, q)
    inversa = implica(q, p)
    contronominale = implica(not q, not p)
    print(f"{str(p):^5}{str(q):^5}{str(originale):^7}{str(inversa):^16}{str(contronominale):^30}")
````

## Esercizio svolto e commentato

**Problema:** considera l'affermazione "se un quadrilatero è un quadrato, allora ha quattro lati uguali". Scrivi la sua negazione corretta, la sua inversa, e la sua contronominale, e valuta la verità di ciascuna.

**Ragionamento:**

Chiamiamo $p$ = "il quadrilatero è un quadrato" e $q$ = "ha quattro lati uguali". L'affermazione originale è $p\Rightarrow q$, ed è **vera** (è proprio una proprietà dei quadrati).

**Negazione** ($p \wedge \neg q$, non "se p allora non q"!): "il quadrilatero è un quadrato e non ha quattro lati uguali". Questa è **falsa**, perché nessun quadrato manca di questa proprietà — coerente con il fatto che l'originale è vera (la negazione di una proposizione vera deve essere falsa).

**Inversa** ($q\Rightarrow p$): "se un quadrilatero ha quattro lati uguali, allora è un quadrato". Questa è **falsa**: un rombo (non quadrato) ha quattro lati uguali ma angoli diversi da 90°, quindi non è un quadrato. Qui vediamo concretamente perché l'inversa non è, in generale, equivalente all'originale: un rombo è un controesempio.

**Contronominale** ($\neg q \Rightarrow \neg p$): "se un quadrilatero non ha quattro lati uguali, allora non è un quadrato". Questa è **vera**, ed è logicamente equivalente all'originale (coerentemente con quanto visto nella teoria): se mancasse anche un solo lato uguale, non potrebbe essere un quadrato, perché tutti i quadrati hanno i lati uguali.

**Perché il rombo è un controesempio così efficace:** un controesempio deve soddisfare l'ipotesi ma non la tesi. Il rombo soddisfa "ha quattro lati uguali" (ipotesi dell'inversa) ma non "è un quadrato" (tesi dell'inversa): è esattamente il caso $p$ falsa, $q$ vera per l'implicazione inversa scritta come $q\Rightarrow p$, quindi... aspetta, controlliamo: qui l'ipotesi dell'inversa è $q$ (lati uguali, vera per il rombo) e la tesi è $p$ (essere quadrato, falsa per il rombo): esattamente l'unico caso che rende falsa un'implicazione, come visto nella tavola di verità.

````python
# Non c'è un calcolo numerico diretto qui (è un esempio geometrico),
# ma possiamo comunque modellare la situazione con proposizioni booleane
# per un caso concreto: il rombo.

quadrato = False       # il rombo non è un quadrato
lati_uguali = True     # il rombo ha quattro lati uguali

p, q = quadrato, lati_uguali
originale = (not p) or q       # p => q
inversa = (not q) or p         # q => p
contronominale = (not (not q)) or (not p)  # (not q) => (not p), cioè q or (not p)

print("Caso concreto: il rombo")
print("p (è quadrato) =", p, "  q (ha lati uguali) =", q)
print("Originale p=>q vera?", originale)
print("Inversa q=>p vera?", inversa)     # deve essere False: il rombo è il controesempio
print("Contronominale vera?", contronominale)
````

## Riepilogo finale

- $p\Rightarrow q$ è falsa **solo** quando $p$ è vera e $q$ è falsa; in tutti gli altri casi (compreso $p$ falsa) è vera.
- La negazione corretta di un'implicazione è $\neg(p\Rightarrow q) \equiv p \wedge \neg q$: serve per costruire controesempi.
- L'**inversa** ($q\Rightarrow p$) non è, in generale, equivalente all'implicazione originale; la **contronominale** ($\neg q \Rightarrow \neg p$) lo è sempre.
- $p\Leftrightarrow q$ ("se e solo se") richiede la verità di entrambe le direzioni $p\Rightarrow q$ e $q\Rightarrow p$; si dimostra sempre nei due versi separatamente.
- Il vocabolario "ipotesi" e "tesi" nasce direttamente dalla struttura dell'implicazione, ed è quello che useremo scrivendo dimostrazioni a partire dal notebook 0.20.
