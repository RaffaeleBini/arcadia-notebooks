## Teoria, passo dopo passo

### Il punto di partenza: ipotesi e tesi

Un teorema ha quasi sempre la forma di un'implicazione (notebook 0.16): $p \Rightarrow q$. La parte $p$ si chiama **ipotesi** (ciò che supponiamo vero), la parte $q$ si chiama **tesi** (ciò che vogliamo dimostrare). Dimostrare il teorema significa costruire una catena di ragionamenti validi che parte dall'ipotesi e arriva alla tesi.

### 1. Dimostrazione diretta

È la strategia più naturale: si parte dall'ipotesi $p$ e, con una sequenza di passaggi ciascuno logicamente giustificato dal precedente, si arriva alla tesi $q$.

*Esempio (già visto, notebook 0.1):* "se $a$ e $b$ sono pari, allora $a+b$ è pari". Partiamo da $a=2k_1$, $b=2k_2$ (ipotesi), e con passaggi algebrici diretti arriviamo a $a+b=2(k_1+k_2)$ (tesi).

### 2. Dimostrazione per contronominale

Come visto nel notebook 0.16, $p\Rightarrow q$ è logicamente equivalente a $(\neg q) \Rightarrow (\neg p)$. A volte è più facile dimostrare quest'ultima che l'originale.

*Esempio:* dimostriamo "se $n^2$ è pari, allora $n$ è pari" (usata implicitamente nel notebook 0.4, per $\sqrt{2}$). Diretta sarebbe scomoda (partire da "$n^2$ pari" e dedurre qualcosa su $n$ non è immediato). Per contronominale, dimostriamo invece "se $n$ è dispari, allora $n^2$ è dispari": scriviamo $n=2k+1$, allora $n^2 = 4k^2+4k+1 = 2(2k^2+2k)+1$, che è dispari (è "il doppio di qualcosa, più 1", proprio come nel notebook 0.1). Questo dimostra la contronominale, quindi anche l'originale.

### 3. Dimostrazione per assurdo

Si suppone **vera l'ipotesi e falsa la tesi** (cioè si suppone falso ciò che si vuole dimostrare) e si mostra che questo porta a una **contraddizione** (un'affermazione palesemente falsa, o due affermazioni tra loro incompatibili). Se supporre il falso porta a un assurdo, allora l'ipotesi di partenza era sbagliata: la tesi deve essere vera.

*Esempio (già visto, notebook 0.4):* per dimostrare che $\sqrt{2}$ è irrazionale, abbiamo supposto il contrario ($\sqrt{2}=\frac{a}{b}$ ridotta ai minimi termini) e siamo arrivati alla contraddizione che sia $a$ sia $b$ dovevano essere pari (contro l'ipotesi di partenza che non avessero fattori comuni).

### 4. Il controesempio (per confutare, non per dimostrare)

Per mostrare che un'affermazione $\forall x \in A,\ p(x)$ è **falsa**, basta esibire un solo $x_0 \in A$ per cui $p(x_0)$ è falsa (regola vista nel notebook 0.17: $\neg(\forall x, p(x)) \equiv \exists x, \neg p(x)$). Un controesempio **non dimostra** nulla di positivo: serve solo a smentire.

*Esempio (già visto, notebook 0.16):* per confutare "se un quadrilatero ha quattro lati uguali, allora è un quadrato", basta il rombo come controesempio.

### 5. Il principio di induzione

Questo è uno schema speciale, che sfrutta direttamente il **buon ordinamento di $\mathbb{N}$** visto nel notebook 0.1. Serve per dimostrare affermazioni della forma $\forall n \in \mathbb{N},\ P(n)$ (una proprietà che deve valere per **tutti** i naturali, infiniti — non possiamo verificarli uno per uno!).

Il principio dice: se riusciamo a dimostrare che

1. **Passo base**: $P(0)$ è vera (o $P(n_0)$, se partiamo da un altro punto di partenza),
2. **Passo induttivo**: per ogni $k$, se $P(k)$ è vera (**ipotesi induttiva**) allora anche $P(k+1)$ è vera,

allora $P(n)$ è vera per **ogni** $n\ge0$.

**Perché funziona (l'intuizione):** pensa a una fila infinita di tessere del domino. Il passo base dice che la prima tessera cade. Il passo induttivo dice che ogni tessera, cadendo, fa cadere la successiva. Insieme, queste due condizioni garantiscono che **tutte** le tessere cadranno, una dopo l'altra, senza doverle spingere una per una a mano. Questo è precisamente ciò che il buon ordinamento di $\mathbb{N}$ rende possibile: non esiste un "primo naturale dove la proprietà smette di valere" senza che il passo induttivo lo contraddica.

````python
import sympy as sp

n = sp.symbols('n', integer=True, nonnegative=True)

# Esempio di induzione: dimostriamo (verifichiamo su vari n) che
# 0 + 1 + 2 + ... + n = n(n+1)/2

def somma_diretta(n):
    return sum(range(n + 1))

def formula(n):
    return n * (n + 1) // 2

print("Verifica della formula 0+1+...+n = n(n+1)/2 per vari n:")
for valore in [0, 1, 5, 10, 50]:
    diretta = somma_diretta(valore)
    da_formula = formula(valore)
    print(f"n={valore}: somma diretta={diretta}, formula={da_formula}, uguali: {diretta == da_formula}")

# Questo NON è una dimostrazione (verifica solo alcuni casi, non tutti gli infiniti n)!
# La vera dimostrazione per induzione è nell'esercizio svolto qui sotto, a parole.
````

## Esercizio svolto e commentato

**Problema:** dimostra per induzione che, per ogni $n \in \mathbb{N}$, vale $0+1+2+\dots+n = \dfrac{n(n+1)}{2}$.

**Ragionamento:**

Chiamiamo $P(n)$ l'affermazione "$0+1+\dots+n = \frac{n(n+1)}{2}$". Seguiamo i due passi del principio di induzione visto nella teoria.

**Passo base — verifichiamo $P(0)$:** la somma da $0$ a $0$ è semplicemente $0$. La formula dà $\frac{0\times(0+1)}{2}=0$. Coincidono: $P(0)$ è vera.

**Passo induttivo — supponiamo vera $P(k)$ (ipotesi induttiva) e dimostriamo $P(k+1)$:**

L'ipotesi induttiva è: $0+1+\dots+k = \dfrac{k(k+1)}{2}$ (la **supponiamo** vera, non la stiamo dimostrando di nuovo — è il "se cade questa tessera" del domino).

Vogliamo dimostrare $P(k+1)$, cioè $0+1+\dots+k+(k+1) = \dfrac{(k+1)(k+2)}{2}$.

Partiamo dal lato sinistro di $P(k+1)$ e usiamo l'ipotesi induttiva per riscrivere la parte $0+1+\dots+k$:

$$0+1+\dots+k+(k+1) = \underbrace{\frac{k(k+1)}{2}}_{\text{per ipotesi induttiva}} + (k+1)$$

Ora facciamo i calcoli algebrici, raccogliendo $(k+1)$ come fattore comune:

$$\frac{k(k+1)}{2}+(k+1) = (k+1)\left(\frac{k}{2}+1\right) = (k+1)\cdot\frac{k+2}{2} = \frac{(k+1)(k+2)}{2}$$

Questo è esattamente il lato destro di $P(k+1)$! Abbiamo quindi dimostrato che, **se** $P(k)$ è vera, **allora** anche $P(k+1)$ lo è.

**Conclusione:** avendo verificato sia il passo base sia il passo induttivo, per il principio di induzione $P(n)$ è vera per ogni $n\in\mathbb{N}$.

**Perché il passo induttivo non è "circolare" (non stiamo dando per scontato ciò che vogliamo dimostrare):** l'ipotesi induttiva riguarda un $k$ **specifico e fissato**, mentre la tesi finale riguarda **tutti** gli $n$. Il passo induttivo dimostra solo l'implicazione "$P(k)\Rightarrow P(k+1)$", non $P(k)$ stessa: è il passo base a garantire che la catena parta da qualcosa di vero, e il passo induttivo a garantire che la verità si propaghi, tessera dopo tessera, senza fine.

````python
def somma_diretta(n):
    return sum(range(n + 1))

def formula(n):
    return n * (n + 1) // 2

# Verifica del passo base
print("Passo base, P(0):", somma_diretta(0), "=", formula(0))

# Verifica del passo induttivo su un k concreto (illustrativo, non è la dimostrazione generale)
k = 7
ipotesi_induttiva = formula(k)          # assumiamo vera la formula per k
verifica_k_piu_1 = ipotesi_induttiva + (k + 1)   # costruiamo P(k+1) a partire da P(k)
atteso_k_piu_1 = formula(k + 1)
print(f"\nCon k={k}: costruendo P(k+1) da P(k) otteniamo {verifica_k_piu_1}, la formula per k+1 dà {atteso_k_piu_1}")
print("Coincidono?", verifica_k_piu_1 == atteso_k_piu_1)
````

Passo induttivo: supponiamo che in ogni insieme di $k$ cavalli siano tutti dello stesso colore; consideriamo un insieme di $k+1$ cavalli, togliamo un cavallo (restano $k$, stesso colore per ipotesi induttiva), poi rimettiamolo e togliamone un altro (restano ancora $k$, stesso colore); quindi tutti i $k+1$ cavalli hanno lo stesso colore.* Dov'è il problema logico?

**1.** Tecnica: **controesempio**. Basta $n=2$: è primo (divisibile solo per 1 e sé stesso) ma è pari, non dispari. Questo confuta l'affermazione universale.

**2.** $P(n)$: "$1+3+\dots+(2n-1)=n^2$". **Passo base** ($n=1$): $1=1^2$. Vero. **Passo induttivo**: supponiamo $1+3+\dots+(2k-1)=k^2$ (ipotesi induttiva). Allora $1+3+\dots+(2k-1)+(2(k+1)-1) = k^2+(2k+1)$. Ma $k^2+2k+1=(k+1)^2$ (prodotto notevole, lo dimostreremo in generale nel notebook 1.2). Quindi $P(k+1)$ è verificata. Per induzione, $P(n)$ vale per ogni $n\ge1$.

**3.** Contronominale: "se $n$ è pari, allora $n^2$ è pari". Scriviamo $n=2k$: allora $n^2=4k^2=2(2k^2)$, che è pari. Questo dimostra la contronominale, quindi anche l'affermazione originale.

**4.** Il problema è nel **passo induttivo**: il ragionamento "togliamo un cavallo, poi lo rimettiamo e ne togliamo un altro" funziona solo se i due sottoinsiemi di $k$ cavalli hanno **almeno un cavallo in comune** (per collegare i due colori tramite quel cavallo condiviso). Questo è vero se $k+1\ge3$ (i due sottoinsiemi da $k$ elementi si sovrappongono), ma **fallisce esattamente nel caso $k=1$** (passando da 1 a 2 cavalli): togliendo un cavallo da un insieme di 2, restano due sottoinsiemi da 1 cavallo ciascuno, che **non hanno nessun elemento in comune** — non c'è un cavallo "ponte" che colleghi i due colori. Il passo induttivo, quindi, non è valido per il primo passaggio ($k=1 \to k=2$), anche se sembra valido per $k$ più grandi: l'intera catena si rompe fin dall'inizio, e la conclusione (falsa, come sappiamo dall'esperienza) non è mai raggiunta legittimamente.

## Riepilogo finale

- Ogni teorema è (quasi sempre) un'implicazione ipotesi $\Rightarrow$ tesi; dimostrarlo significa collegare le due con una catena di passaggi logicamente validi.
- **Dimostrazione diretta**: dall'ipotesi alla tesi, passo dopo passo. **Per contronominale**: si dimostra $\neg q \Rightarrow \neg p$ (equivalente all'originale). **Per assurdo**: si suppone falsa la tesi e si arriva a una contraddizione.
- Un **controesempio** confuta un'affermazione universale, ma non dimostra nulla in positivo: basta un solo caso che viola la proprietà.
- Il **principio di induzione** dimostra proprietà per tutti i naturali con due passi: base ($P(0)$ vera) e induttivo ($P(k)\Rightarrow P(k+1)$ per ogni $k$) — si basa sul buon ordinamento di $\mathbb{N}$ visto nel notebook 0.1.
- Conoscere questi schemi ti aiuterà a riconoscerli in ogni dimostrazione che incontrerai da qui fino alla fine del corso, e a costruirne di tue con più sicurezza.
