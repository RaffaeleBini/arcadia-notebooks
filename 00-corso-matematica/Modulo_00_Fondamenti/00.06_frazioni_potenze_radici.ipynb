## Perché ci interessa

Ora che abbiamo tutti gli insiemi numerici principali ($\mathbb{N}, \mathbb{Z}, \mathbb{Q}, \mathbb{R}$), dobbiamo consolidare gli strumenti di calcolo che useremo **continuamente** da qui in avanti: le potenze e le radici.

Le potenze nascono per scrivere in modo compatto le moltiplicazioni ripetute ($2\times2\times2 = 2^3$), e le radici nascono come loro "operazione inversa": rispondono alla domanda "quale numero, moltiplicato per sé stesso $n$ volte, dà questo risultato?". Sono strumenti che incontrerai in ogni singolo modulo successivo, quindi vale la pena fissarne bene le regole ora.

## Teoria, passo dopo passo

### Potenze con esponente naturale

Per $n$ naturale positivo, la potenza $a^n$ è la moltiplicazione di $a$ per sé stesso $n$ volte:

$$a^n = \underbrace{a \times a \times \dots \times a}_{n \text{ volte}}$$

Da questa definizione derivano le proprietà delle potenze, che **non sono regole arbitrarie** ma conseguenze dirette di "quante volte moltiplico":

$$a^m \times a^n = a^{m+n} \qquad \text{(sommo le occorrenze di } a\text{)}$$
$$\frac{a^m}{a^n} = a^{m-n} \qquad \text{(le occorrenze si "elidono")}$$
$$(a^m)^n = a^{mn} \qquad \text{(ripeto } n \text{ volte un blocco di } m \text{ occorrenze)}$$

### Esponente zero e negativo: estendere la definizione con coerenza

Cosa significa $a^0$? Usiamo la regola $\frac{a^m}{a^n}=a^{m-n}$ con $m=n$: $\frac{a^n}{a^n} = a^{n-n} = a^0$. Ma $\frac{a^n}{a^n}=1$ (qualsiasi numero diviso per sé stesso, tranne 0). Quindi, per coerenza, definiamo:

$$a^0 = 1 \qquad (a \ne 0)$$

Allo stesso modo, cosa significa $a^{-n}$? Usiamo ancora la stessa regola: $a^0 \div a^n = a^{0-n} = a^{-n}$, cioè $\frac{1}{a^n} = a^{-n}$. Definiamo quindi:

$$a^{-n} = \frac{1}{a^n} \qquad (a \ne 0)$$

Non stiamo "inventando" queste regole: le stiamo **scegliendo in modo che le proprietà delle potenze continuino a valere** anche fuori dal caso degli esponenti naturali positivi. È lo stesso spirito che abbiamo usato nel notebook 0.2 per giustificare "meno per meno fa più".

### Le radici come operazione inversa

La radice $n$-esima di $a$, scritta $\sqrt[n]{a}$, è il numero che elevato alla $n$ dà $a$:

$$\sqrt[n]{a} = b \iff b^n = a$$

Per esempio $\sqrt[3]{8}=2$ perché $2^3=8$. Un caso speciale importante: $\sqrt{a}$ (senza indice, sottinteso $n=2$) è la radice quadrata.

**Attenzione a un errore comune**: $\sqrt{a^2} = |a|$, non semplicemente $a$. Per esempio $\sqrt{(-3)^2} = \sqrt{9} = 3$, non $-3$. Questo perché la radice quadrata, per essere una funzione ben definita, restituisce sempre il risultato **non negativo** (approfondiremo il valore assoluto nel notebook 0.8).

### Il collegamento tra radici e potenze con esponente frazionario

Le radici si possono scrivere anche come potenze con esponente frazionario:

$$\sqrt[n]{a} = a^{1/n}$$

Perché questa scelta è coerente? Verifichiamo con la proprietà $(a^m)^n = a^{mn}$: se eleviamo $a^{1/n}$ alla $n$, otteniamo $\left(a^{1/n}\right)^n = a^{(1/n)\times n} = a^1 = a$. Esattamente la proprietà che definisce la radice $n$-esima! Questo ci permette di trattare radici e potenze con le stesse regole, il che sarà molto comodo nei calcoli futuri (per esempio: $\sqrt[3]{a^2} = a^{2/3}$).

````python
import sympy as sp

a = sp.symbols('a', positive=True)  # positive=True evita ambiguità con radici e valori assoluti

# Verifichiamo le proprietà delle potenze con SymPy
espressione1 = sp.simplify(a**3 * a**5 - a**8)
print("a^3 * a^5 - a^8 =", espressione1, "(deve fare 0: le proprietà coincidono)")

espressione2 = sp.simplify((a**2)**3 - a**6)
print("(a^2)^3 - a^6 =", espressione2)

# Radice quadrata di un quadrato: attenzione al valore assoluto
x = sp.symbols('x', real=True)  # x reale qualsiasi, non necessariamente positivo
radice_di_quadrato = sp.sqrt(x**2)
print("\nsqrt(x^2) semplificato da SymPy:", sp.simplify(radice_di_quadrato))  # deve dare Abs(x)

# Radice come potenza frazionaria
radice_cubica = sp.root(8, 3)
potenza_frazionaria = sp.Rational(8) ** sp.Rational(1, 3)
print(f"\ncubica di 8 = {radice_cubica}, 8^(1/3) = {potenza_frazionaria}")
````

## Esercizio svolto e commentato

**Problema:** semplifica $\dfrac{a^5 \cdot a^{-2}}{a^3} \cdot \sqrt[3]{a^6}$ (con $a>0$).

**Ragionamento:**

Lavoriamo prima sulla frazione, usando le proprietà delle potenze viste sopra. Al numeratore, sommiamo gli esponenti (stessa base, moltiplicazione):

$$a^5 \cdot a^{-2} = a^{5+(-2)} = a^3$$

Ora dividiamo per $a^3$: sottraiamo gli esponenti (stessa base, divisione):

$$\frac{a^3}{a^3} = a^{3-3} = a^0 = 1$$

Quindi la frazione si semplifica completamente a $1$. Restiamo con:

$$1 \cdot \sqrt[3]{a^6}$$

Riscriviamo la radice come potenza frazionaria: $\sqrt[3]{a^6} = a^{6/3} = a^2$.

Il risultato finale è:

$$a^2$$

**Perché conviene riscrivere le radici come potenze:** con la notazione $a^{p/q}$ possiamo applicare le stesse regole (somma/sottrazione di esponenti) usate per le potenze intere, invece di dover ricordare regole separate per le radici. È un esempio di come una buona notazione **riduca** il numero di cose da imparare a memoria.

````python
import sympy as sp

a = sp.symbols('a', positive=True)
espressione = (a**5 * a**-2 / a**3) * sp.root(a**6, 3)
semplificata = sp.simplify(espressione)
print("Espressione semplificata:", semplificata)  # atteso a^2
````

## Riepilogo finale

- $a^n$ conta quante volte moltiplichi $a$ per sé stesso; le proprietà delle potenze (somma/sottrazione di esponenti) derivano direttamente da questa idea.
- $a^0=1$ e $a^{-n}=\frac{1}{a^n}$ non sono regole arbitrarie: sono le uniche scelte che mantengono coerenti le proprietà delle potenze anche fuori dal caso degli esponenti naturali.
- La radice $n$-esima è l'operazione inversa della potenza $n$-esima: $\sqrt[n]{a}=a^{1/n}$.
- $\sqrt{a^2}=|a|$, non $a$: la radice quadrata restituisce sempre un valore non negativo.
- Scrivere le radici come potenze frazionarie permette di usare un'unica famiglia di regole di calcolo per entrambe.
