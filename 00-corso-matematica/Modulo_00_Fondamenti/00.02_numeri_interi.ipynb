# 0.2 — Numeri interi

## Perché ci interessa

Nel notebook precedente abbiamo scoperto un limite dei numeri naturali: la sottrazione non è sempre possibile. $3 - 5$ non ha risposta in $\mathbb{N}$.

Ma nella vita reale, questa domanda ha senso eccome: se ho 3 euro e ne devo 5, sono "indietro" di 2 euro. Se la temperatura è di 3 gradi e scende di 5 gradi, arriva a "meno 2" gradi. Ci serve un modo per rappresentare quantità **al di sotto dello zero**.

L'idea è semplice: allarghiamo l'insieme dei numeri aggiungendo, per ogni naturale positivo, il suo "opposto" negativo. Questo nuovo insieme si chiama insieme dei **numeri interi**.

## Teoria, passo dopo passo

### Costruire i numeri negativi

Pensa a un termometro, o meglio a una retta graduata come quella di un termometro: 0 al centro, i numeri positivi salgono verso destra, e ora aggiungiamo dei numeri che scendono verso sinistra: $-1, -2, -3, \dots$

Il numero $-n$ (con $n$ naturale positivo) è definito come **l'opposto di $n$**: l'unico numero che, sommato a $n$, dà 0.

$$n + (-n) = 0$$

Questa è la proprietà che definisce l'opposto, ed è anche il motivo per cui li introduciamo: ci serve un numero che "annulli" $n$ nella somma, così da poter sempre risolvere equazioni come $x + 5 = 3$ (che in $\mathbb{N}$ non aveva soluzione).

### L'insieme dei numeri interi

$$\mathbb{Z} = \{\dots, -3, -2, -1, 0, 1, 2, 3, \dots\}$$

(la lettera $\mathbb{Z}$ viene dal tedesco *Zahlen*, "numeri"). Notiamo che $\mathbb{N} \subset \mathbb{Z}$: ogni naturale è anche un intero (i naturali sono gli interi "non negativi").

### Ora la sottrazione funziona sempre

Definiamo la sottrazione a partire dall'addizione e dall'opposto:

$$a - b := a + (-b)$$

Con questa definizione, $3 - 5 = 3 + (-5) = -2$: ha finalmente una risposta! $\mathbb{Z}$ è **chiuso** rispetto a addizione, moltiplicazione e ora anche sottrazione.

### La moltiplicazione con i segni

Una regola che si impara spesso a memoria, ma che vale la pena capire: perché "meno per meno fa più"?

Pensiamo alla moltiplicazione come "ripetizione" o "opposizione". Sappiamo che:

- $(+) \times (+) = (+)$: ripetere un guadagno resta un guadagno. Es. $2 \times 3 = 6$.
- $(+) \times (-) = (-)$: ripetere un debito resta un debito. Es. $2 \times (-3) = -6$ (due debiti di 3 fanno un debito di 6).
- $(-) \times (+) = (-)$: per la proprietà commutativa, è come sopra. $(-3) \times 2 = -6$.
- $(-) \times (-) = (+)$: questo è il caso che sembra strano. Vediamolo con la coerenza algebrica: vogliamo che valga la proprietà distributiva anche con i negativi. Consideriamo:

$$(-2) \times \big(3 + (-3)\big) = (-2) \times 0 = 0$$

Ma per la proprietà distributiva questo deve essere anche uguale a:

$$(-2)\times 3 + (-2)\times(-3) = -6 + (-2)\times(-3)$$

Perché la somma faccia 0, deve essere $(-2)\times(-3) = 6$. Non è un capriccio: è l'unica scelta che mantiene coerenti le regole (distributiva, opposto) che già valevano in $\mathbb{N}$. Se scegliessimo un'altra regola, l'algebra "si romperebbe".

### Cosa manca ancora

Anche $\mathbb{Z}$ ha un limite: la divisione non è sempre possibile. $7 \div 2$ non è un numero intero. Questo sarà il motivo per passare, nel prossimo notebook, ai numeri razionali.

````python
# Verifichiamo con Python alcune delle proprietà viste.

# 1. La sottrazione ora è sempre definita e produce un intero (anche negativo)
a, b = 3, 5
print(f"{a} - {b} = {a - b}")  # -2, un intero valido

# 2. Verifichiamo la regola "meno per meno fa più" tramite la coerenza distributiva
import itertools
for x, y in itertools.product([-4, -1, 0, 2, 5], repeat=2):
    # controlliamo che x * (y + (-y)) = 0 sia sempre coerente con la distributiva
    lhs = x * (y + (-y))
    rhs = x * y + x * (-y)
    assert lhs == 0 == rhs, "la coerenza algebrica deve valere sempre"
print("Coerenza della moltiplicazione con i segni verificata su vari esempi.")

# 3. N è sottoinsieme di Z: ogni naturale è anche intero
naturali = list(range(0, 6))
interi_corrispondenti = [n for n in naturali]  # stessa rappresentazione, nessuna perdita
print("Naturali:", naturali)
print("Sono anche interi:", interi_corrispondenti)
````

## Esercizio svolto e commentato

**Problema:** dimostra che l'opposto della somma di due interi è la somma degli opposti, cioè $-(a+b) = (-a) + (-b)$.

**Ragionamento:**

Per dimostrare che due espressioni sono uguali, un metodo efficace è mostrare che sommate a $(a+b)$ danno entrambe 0 — perché l'opposto di un numero è **unico** (c'è un solo numero che, sommato a $(a+b)$, dà 0).

Calcoliamo $(a+b) + \big((-a)+(-b)\big)$, riorganizzando i termini con le proprietà associativa e commutativa:

$$(a+b) + \big((-a)+(-b)\big) = \big(a + (-a)\big) + \big(b + (-b)\big) = 0 + 0 = 0$$

Quindi $(-a)+(-b)$ è un numero che, sommato ad $(a+b)$, dà 0. Ma per definizione l'opposto di $(a+b)$, cioè $-(a+b)$, è *l'unico* numero con questa proprietà. Dunque devono coincidere:

$$-(a+b) = (-a)+(-b)$$

**Perché "riorganizzare i termini" è lecito:** stiamo usando solo due proprietà che sappiamo già valere in $\mathbb{Z}$ (associativa e commutativa dell'addizione), non stiamo inventando nulla di nuovo. Questo è tipico delle dimostrazioni algebriche: si combinano proprietà già note per arrivarne a una nuova.

````python
# Verifica numerica dell'esercizio
casi = [(3, 7), (-4, 9), (-2, -5), (0, 6)]
for a, b in casi:
    sinistra = -(a + b)
    destra = (-a) + (-b)
    print(f"-({a}+{b}) = {sinistra}   (-{a})+(-{b}) = {destra}   uguali: {sinistra == destra}")
````

## Riepilogo finale

- I numeri interi $\mathbb{Z} = \{\dots,-2,-1,0,1,2,\dots\}$ estendono $\mathbb{N}$ aggiungendo gli opposti dei naturali positivi.
- Ogni intero $n$ ha un opposto $-n$ tale che $n + (-n) = 0$; la sottrazione si definisce come $a - b = a + (-b)$.
- $\mathbb{Z}$ è chiuso rispetto ad addizione, sottrazione e moltiplicazione (ma non rispetto alla divisione).
- La regola "meno per meno fa più" non è arbitraria: è l'unica scelta compatibile con le proprietà algebriche già valide in $\mathbb{N}$ (in particolare la distributiva).
- $\mathbb{Z}$ non ha un elemento minimo assoluto: il buon ordinamento di $\mathbb{N}$ non si estende a tutto $\mathbb{Z}$.
