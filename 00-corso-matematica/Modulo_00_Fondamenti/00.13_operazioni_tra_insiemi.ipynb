# 0.13 — Operazioni tra insiemi

**Prerequisiti:** 0.12 Insiemi.


## Perché ci interessa

Nel notebook precedente abbiamo imparato cosa sono gli insiemi e come confrontarli (appartenenza, inclusione, uguaglianza). Ora ci serve un modo per **combinarli**: prendere due insiemi e costruirne uno nuovo.

Questo ci servirà moltissimo più avanti: per esempio, il dominio di una funzione con più condizioni (Modulo 3) si ottiene spesso intersecando gli insiemi dove valgono le singole condizioni. Le operazioni che vedremo qui (unione, intersezione, differenza, complementare) sono gli "attrezzi" di base per lavorare con gli insiemi.

## Teoria, passo dopo passo

Immaginiamo di avere due insiemi, per esempio $A=\{1,2,3,4\}$ e $B=\{3,4,5,6\}$, e chiediamoci: in quanti modi diversi possiamo "combinarli"?

### Unione

L'**unione** di $A$ e $B$, scritta $A \cup B$, è l'insieme di tutti gli elementi che stanno in $A$ **oppure** in $B$ (oppure in entrambi):

$$A \cup B = \{x : x \in A \text{ oppure } x \in B\}$$

Con il nostro esempio: $A \cup B = \{1,2,3,4,5,6\}$ (ogni elemento che compare in almeno uno dei due).

### Intersezione

L'**intersezione** di $A$ e $B$, scritta $A \cap B$, è l'insieme degli elementi che stanno **sia** in $A$ **sia** in $B$ contemporaneamente:

$$A \cap B = \{x : x \in A \text{ e } x \in B\}$$

Con il nostro esempio: $A \cap B = \{3,4\}$ (solo gli elementi comuni a entrambi).

Se $A \cap B = \emptyset$ (nessun elemento in comune), diciamo che $A$ e $B$ sono **disgiunti**.

### Differenza

La **differenza** $A \setminus B$ (si legge "$A$ meno $B$") è l'insieme degli elementi di $A$ che **non** stanno in $B$:

$$A \setminus B = \{x : x \in A \text{ e } x \notin B\}$$

Con il nostro esempio: $A \setminus B = \{1,2\}$ (gli elementi di $A$ che non sono anche in $B$). Nota che $A\setminus B$ in generale è diverso da $B\setminus A$: qui $B\setminus A=\{5,6\}$. La differenza **non è commutativa**.

### Complementare

Quando lavoriamo dentro un **insieme universo** $U$ (l'insieme "di riferimento" che contiene tutti gli elementi di cui ci occupiamo, per esempio $U=\mathbb{N}$), il **complementare** di $A$ rispetto a $U$, scritto $A^c$ o $\overline{A}$, è l'insieme degli elementi di $U$ che non stanno in $A$:

$$A^c = U \setminus A = \{x \in U : x \notin A\}$$

Esempio: se $U = \{1,2,\dots,10\}$ e $A=\{2,4,6,8,10\}$ (i pari), allora $A^c=\{1,3,5,7,9\}$ (i dispari): il complementare dipende sempre da quale universo scegliamo.

### Le leggi di De Morgan

Un risultato importante, che userete spesso (anche in logica, notebook 0.15-0.16): il complementare di un'unione è l'intersezione dei complementari, e viceversa:

$$(A \cup B)^c = A^c \cap B^c \qquad (A \cap B)^c = A^c \cup B^c$$

Intuitivamente: "non essere in $A$ o in $B$" equivale a "non essere in $A$ **e** non essere in $B$" (prima legge); "non essere sia in $A$ sia in $B$" equivale a "non essere in $A$ **oppure** non essere in $B$" (seconda legge). Verificheremo queste leggi anche visivamente con i diagrammi di Venn nella cella di codice qui sotto.

````python
import matplotlib.pyplot as plt

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("A ∪ B =", A | B)   # unione
print("A ∩ B =", A & B)   # intersezione
print("A \\ B =", A - B)  # differenza
print("B \\ A =", B - A)  # differenza (nell'altro verso, per mostrare che non è commutativa)

# Complementare rispetto a un universo U
U = set(range(1, 11))
A2 = {2, 4, 6, 8, 10}
complementare_A2 = U - A2
print("\nUniverso U =", U)
print("A2 =", A2)
print("Complementare di A2 rispetto a U =", complementare_A2)

# Verifica delle leggi di De Morgan
sinistra1 = U - (A | B if (A | B).issubset(U) else set())  # esempio semplificato, vedi sotto per un caso pulito
# Rifacciamo con A, B sottoinsiemi dello stesso universo U per un confronto corretto
U2 = set(range(1, 8))
A3, B3 = {1, 2, 3}, {3, 4, 5}
sinistra_demorgan1 = U2 - (A3 | B3)
destra_demorgan1 = (U2 - A3) & (U2 - B3)
print("\n(A∪B)^c =", sinistra_demorgan1)
print("A^c ∩ B^c =", destra_demorgan1)
print("Coincidono?", sinistra_demorgan1 == destra_demorgan1)

# Diagramma di Venn "fatto a mano" con due cerchi sovrapposti (solo Matplotlib)
fig, ax = plt.subplots(figsize=(5, 5))
cerchio_A = plt.Circle((-0.6, 0), 1.5, alpha=0.4, color='tab:blue', label='A')
cerchio_B = plt.Circle((0.6, 0), 1.5, alpha=0.4, color='tab:orange', label='B')
ax.add_patch(cerchio_A)
ax.add_patch(cerchio_B)
ax.text(-1.3, 0, "A", fontsize=14)
ax.text(1.3, 0, "B", fontsize=14)
ax.text(0, 0, "A∩B", fontsize=11, ha='center')
ax.set_xlim(-3, 3)
ax.set_ylim(-2, 2)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title("A ∪ B = tutta l'area colorata; A ∩ B = zona di sovrapposizione")
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dati $A=\{n\in\mathbb{N} : n \le 15,\ n \text{ pari}\}$ e $B=\{n\in\mathbb{N}: n\le15,\ n \text{ multiplo di } 3\}$, trova $A\cap B$ e spiega il ragionamento.

**Ragionamento:**

Prima elenchiamo entrambi gli insiemi, perché sono finiti e piccoli, così possiamo lavorare per confronto diretto (per insiemi più grandi o infiniti servirebbe un ragionamento più astratto, come vedremo più avanti nel corso).

$A$ = naturali pari fino a 15: $\{0,2,4,6,8,10,12,14\}$.

$B$ = naturali multipli di 3 fino a 15: $\{0,3,6,9,12,15\}$.

Per trovare $A\cap B$ dobbiamo cercare gli elementi che compaiono in **entrambe** le liste. Scorriamo $A$ e controlliamo, uno per uno, quali suoi elementi sono anche in $B$:

- $0$: è in $B$? Sì (0 è multiplo di ogni numero). ✓
- $2$: è in $B$? No.
- $4$: No.
- $6$: Sì. ✓
- $8$: No.
- $10$: No.
- $12$: Sì. ✓
- $14$: No.

Quindi $A\cap B = \{0,6,12\}$.

**Osservazione più profonda:** questi sono esattamente i multipli di $6$ fino a 15 (perché essere sia pari sia multiplo di 3 equivale, per un numero, a essere multiplo del minimo comune multiplo tra 2 e 3, cioè 6 — un'idea che ritroveremo parlando di aritmetica modulare nel Modulo 18, ma che si può già intuire qui).

**Perché elencare esplicitamente, in questo caso, è un metodo valido:** quando gli insiemi sono finiti e piccoli, il confronto diretto elemento per elemento è rigoroso quanto un ragionamento più astratto — non è "meno matematico", è semplicemente il metodo più diretto disponibile per questo caso specifico.

````python
A = {n for n in range(0, 16) if n % 2 == 0}
B = {n for n in range(0, 16) if n % 3 == 0}

print("A =", sorted(A))
print("B =", sorted(B))
print("A ∩ B =", sorted(A & B))

# Confronto con i multipli di 6
multipli_di_6 = {n for n in range(0, 16) if n % 6 == 0}
print("\nMultipli di 6 fino a 15:", sorted(multipli_di_6))
print("Coincide con A ∩ B?", (A & B) == multipli_di_6)
````

## Riepilogo finale

- **Unione** $A\cup B$: elementi in $A$ oppure in $B$ (o entrambi). **Intersezione** $A\cap B$: elementi in entrambi contemporaneamente.
- **Differenza** $A\setminus B$: elementi di $A$ che non sono in $B$; non è commutativa ($A\setminus B \ne B\setminus A$ in generale).
- Il **complementare** $A^c$ dipende sempre da un insieme universo $U$ di riferimento: $A^c = U\setminus A$.
- Le **leggi di De Morgan** collegano unione, intersezione e complementare: negare un "oppure" dà un "e", e viceversa. Le ritroveremo identiche in logica (notebook 0.15-0.16).
