# 0.17 — Quantificatori ∀ e ∃

**Prerequisiti:** 0.15 Logica elementare, 0.16 Implicazione, equivalenza, negazione, 0.12 Insiemi.


## Perché ci interessa

Nel notebook 0.15 abbiamo detto che "$x+1=5$" non è una proposizione, perché il suo valore di verità dipende da $x$: è vera se $x=4$, falsa altrimenti. Espressioni come questa, che contengono una variabile, si chiamano **predicati**.

Come trasformiamo un predicato in una vera proposizione (con valore di verità fissato)? Specificando **quanti** valori della variabile devono soddisfarlo: "tutti" oppure "almeno uno". Questa è l'idea dei **quantificatori**, uno strumento che troverai scritto in ogni definizione rigorosa e in ogni teorema da qui fino alla fine del corso.

## Teoria, passo dopo passo

### Il quantificatore universale ∀

Il simbolo $\forall$ (una A capovolta, da "All", tutti) si legge "per ogni". La proposizione

$$\forall x \in A,\ p(x)$$

si legge "per ogni $x$ appartenente ad $A$, vale $p(x)$", ed è **vera** se e solo se **tutti** gli elementi di $A$ soddisfano il predicato $p$.

Esempio: $\forall n \in \mathbb{N},\ n^2 \ge 0$ è vera (ogni naturale al quadrato è non negativo). Esempio falso: $\forall n \in \mathbb{N},\ n > 5$ è falsa (basta $n=0$ per smentirla).

### Il quantificatore esistenziale ∃

Il simbolo $\exists$ (una E rovesciata, da "Exists") si legge "esiste". La proposizione

$$\exists x \in A,\ p(x)$$

si legge "esiste (almeno) un $x$ appartenente ad $A$ tale che vale $p(x)$", ed è **vera** se e solo se **almeno un** elemento di $A$ soddisfa il predicato $p$ (non serve che sia unico).

Esempio: $\exists n \in \mathbb{N},\ n^2 = 9$ è vera (basta $n=3$). Un simbolo aggiuntivo, $\exists!$, significa "esiste un unico": $\exists! n\in\mathbb{N}, n^2=9$ è anch'essa vera (solo $n=3$ funziona, tra i naturali).

### Come si negano i quantificatori

Questa è la parte più delicata, e la più utile per scrivere controesempi. La regola è:

$$\neg\big(\forall x \in A,\ p(x)\big) \quad \equiv \quad \exists x \in A,\ \neg p(x)$$

"Non è vero che tutti soddisfano $p$" equivale a "esiste (almeno) un elemento che non soddisfa $p$". Ha senso: per smentire un "per tutti", basta un solo controesempio (esattamente come abbiamo visto nel notebook 0.16 per l'inversa di un'implicazione — un rombo bastava a smentire "lati uguali $\Rightarrow$ quadrato").

Analogamente:

$$\neg\big(\exists x \in A,\ p(x)\big) \quad \equiv \quad \forall x \in A,\ \neg p(x)$$

"Non esiste nessuno che soddisfi $p$" equivale a "tutti non soddisfano $p$". Il quantificatore **cambia tipo** quando si nega ($\forall \leftrightarrow \exists$), e il predicato interno si nega a sua volta. Questa regola si chiama a volte "legge di De Morgan per i quantificatori", per l'evidente parallelismo con quanto già visto per insiemi e logica proposizionale.

### Attenzione all'ordine dei quantificatori

Quando compaiono più quantificatori insieme, **l'ordine conta moltissimo** e cambiarlo può cambiare completamente il significato. Confronta:

$$\forall x \in \mathbb{R},\ \exists y \in \mathbb{R},\ y > x$$

"Per ogni $x$, esiste un $y$ più grande di $x$" — vera (basta prendere $y=x+1$, che dipende da $x$).

$$\exists y \in \mathbb{R},\ \forall x \in \mathbb{R},\ y > x$$

"Esiste un $y$ che è più grande di ogni $x$" — falsa (non esiste un numero reale più grande di *tutti* gli altri; per qualunque $y$ scelto, basta prendere $x=y+1$ per smentirlo). Stesse parole, stessi simboli, ordine diverso, significato completamente diverso: quando incontrerai definizioni con quantificatori multipli (per esempio la definizione di limite, nel Modulo 5), l'ordine sarà sempre cruciale.

````python
# Verifichiamo alcuni esempi di proposizioni quantificate direttamente in Python,
# usando all() per il "per ogni" ed any() per "esiste"

naturali_piccoli = range(0, 20)  # un campione finito per la verifica (N è infinito!)

# ∀n, n^2 >= 0
proposizione1 = all(n**2 >= 0 for n in naturali_piccoli)
print("∀n∈N, n^2>=0 :", proposizione1)

# ∀n, n > 5  (falsa: basta n=0)
proposizione2 = all(n > 5 for n in naturali_piccoli)
print("∀n∈N, n>5   :", proposizione2)

# ∃n, n^2 == 9
proposizione3 = any(n**2 == 9 for n in naturali_piccoli)
print("∃n∈N, n^2=9 :", proposizione3)

# Negazione di un "per ogni": deve coincidere con "esiste un controesempio"
negazione_diretta = not proposizione2
esiste_controesempio = any(not (n > 5) for n in naturali_piccoli)
print("\nnot(∀n,n>5) :", negazione_diretta)
print("∃n, not(n>5):", esiste_controesempio)
print("Coincidono? ", negazione_diretta == esiste_controesempio)
````

## Esercizio svolto e commentato

**Problema:** nega la proposizione $\forall x \in \mathbb{R},\ (x>0 \Rightarrow x^2>0)$.

**Ragionamento:**

Applichiamo le regole viste, un passo alla volta, dall'esterno verso l'interno.

**Passo 1 — negare il $\forall$:** per la regola dei quantificatori, negare "per ogni $x$, vale $P(x)$" dà "esiste un $x$ tale che non vale $P(x)$", dove qui $P(x)$ è l'intera implicazione $x>0\Rightarrow x^2>0$:

$$\exists x \in \mathbb{R},\ \neg\big(x>0 \Rightarrow x^2>0\big)$$

**Passo 2 — negare l'implicazione interna:** usando la regola vista nel notebook 0.16, $\neg(p\Rightarrow q) \equiv p \wedge \neg q$, con $p$="$x>0$" e $q$="$x^2>0$":

$$\exists x \in \mathbb{R},\ \big(x>0 \ \wedge\ x^2 \le 0\big)$$

(abbiamo scritto $\neg(x^2>0)$ come $x^2 \le 0$, la negazione diretta di "maggiore di").

Questa è la negazione corretta: "esiste un numero reale positivo il cui quadrato non è positivo". È interessante notare che questa negazione è **falsa** (nessun numero reale positivo ha quadrato non positivo): questo ci conferma, indirettamente, che la proposizione originale era **vera**.

**Perché procedere "un passo alla volta, dall'esterno" evita errori:** negare tutta l'espressione in un colpo solo porta quasi sempre a errori (dimenticare di cambiare il quantificatore, o di negare correttamente l'implicazione). Applicando una regola alla volta, come in una ricetta, il risultato è verificabile a ogni passaggio.

````python
# Verifichiamo su un campione di numeri reali (approssimato con float) che
# la proposizione originale sia vera e la sua negazione sia falsa

import numpy as np

campione = np.linspace(-5, 5, 1000)

originale = all((x <= 0) or (x**2 > 0) for x in campione)  # x>0 => x^2>0, riscritta come (not x>0) or (x^2>0)
negazione = any((x > 0) and (x**2 <= 0) for x in campione)

print("Proposizione originale vera (su campione)?", originale)
print("Sua negazione vera (su campione)?", negazione)
````

## Riepilogo finale

- $\forall x \in A,\ p(x)$ ("per ogni") è vera se **tutti** gli elementi di $A$ soddisfano $p$; $\exists x \in A,\ p(x)$ ("esiste") è vera se **almeno uno** la soddisfa.
- Negare un quantificatore lo scambia con l'altro e nega il predicato interno: $\neg\forall \equiv \exists\neg$, e $\neg\exists \equiv \forall\neg$.
- Per smentire un "per ogni" basta un solo controesempio; per smentire un "esiste" bisogna mostrare che nessun elemento funziona.
- L'ordine dei quantificatori multipli cambia il significato: $\forall x\,\exists y$ e $\exists y\,\forall x$ sono, in generale, proposizioni diverse.
- Questi simboli sono il linguaggio con cui, da qui in avanti, scriveremo definizioni rigorose (per esempio la definizione di limite, nel Modulo 5).
