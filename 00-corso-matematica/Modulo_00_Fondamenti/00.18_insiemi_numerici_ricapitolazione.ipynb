# 0.18 — Insiemi numerici (ricapitolazione N, Z, Q, R)

## Perché ci interessa

Abbiamo costruito, un passo alla volta, quattro insiemi numerici: $\mathbb{N}$, $\mathbb{Z}$, $\mathbb{Q}$, $\mathbb{R}$. Ogni volta, il passaggio all'insieme successivo è nato da un **limite** dell'insieme precedente: la sottrazione non sempre possibile in $\mathbb{N}$, la divisione non sempre possibile in $\mathbb{Z}$, l'esistenza di "buchi" nella retta di $\mathbb{Q}$.

Prima di proseguire con relazioni e funzioni, vale la pena fermarsi e vedere il quadro d'insieme: come si incastrano questi insiemi, quali operazioni sono garantite in ciascuno, e perché questa progressione non è stata casuale.

## Teoria, passo dopo passo

### La catena di inclusioni

$$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}$$

Ogni insieme è un sottoinsieme proprio del successivo (nel senso del notebook 0.12: ogni elemento del più piccolo è anche elemento del più grande, ma non viceversa). Vediamo un elemento che sta in ciascun insieme "aggiuntivo" ma non nel precedente:

- $-3 \in \mathbb{Z}$ ma $-3 \notin \mathbb{N}$.
- $\frac{1}{2} \in \mathbb{Q}$ ma $\frac{1}{2} \notin \mathbb{Z}$.
- $\sqrt{2} \in \mathbb{R}$ ma $\sqrt{2} \notin \mathbb{Q}$.

### Tabella riassuntiva: quali operazioni sono sempre possibili

| Insieme | Addizione | Sottrazione | Moltiplicazione | Divisione (per non-zero) | "Buchi" sulla retta? |
|---|---|---|---|---|---|
| $\mathbb{N}$ | ✓ | ✗ | ✓ | ✗ | (discreto, non ha senso parlare di buchi) |
| $\mathbb{Z}$ | ✓ | ✓ | ✓ | ✗ | (discreto) |
| $\mathbb{Q}$ | ✓ | ✓ | ✓ | ✓ | Sì (mancano gli irrazionali) |
| $\mathbb{R}$ | ✓ | ✓ | ✓ | ✓ | No (completo) |

Ogni "✗" nella tabella è stato, storicamente, il motivo per costruire l'insieme successivo: la matematica non ha inventato numeri nuovi per capriccio, ma per rispondere a domande che l'insieme precedente lasciava senza risposta.

### Perché ci siamo fermati a R (per ora)

Potrebbe sembrare che il processo di allargamento debba continuare all'infinito. In un certo senso è così: nel Modulo 9 vedremo i **numeri complessi** $\mathbb{C}$, che risolvono un altro limite di $\mathbb{R}$ (l'equazione $x^2=-1$ non ha soluzioni reali). Ma per costruire l'analisi matematica (limiti, derivate, integrali — Moduli 4, 5, 7...), $\mathbb{R}$ è l'insieme giusto: la sua proprietà di completezza (nessun "buco") è precisamente ciò che serve per far funzionare i concetti di limite e continuità, come vedremo con rigore nel Modulo 5.

### Densità vs completezza: due proprietà diverse da non confondere

Un punto sottile, utile da avere chiaro: $\mathbb{Q}$ è **denso** in $\mathbb{R}$ (tra due reali distinti c'è sempre un razionale, e viceversa — visto nel notebook 0.3), eppure $\mathbb{Q}$ **non è completo** (ha "buchi", come $\sqrt{2}$). Densità e completezza sono proprietà diverse: la densità dice che i razionali sono "sparsi ovunque", la completezza dice che non manca "nessun punto" alla retta. $\mathbb{Q}$ ha la prima proprietà ma non la seconda; $\mathbb{R}$ le ha entrambe.

````python
import sympy as sp

# Verifichiamo l'appartenenza di alcuni numeri ai vari insiemi con SymPy
numeri = [-3, sp.Rational(1, 2), sp.sqrt(2), 7, sp.pi]

for n in numeri:
    è_naturale = n.is_integer and n >= 0 if n.is_number and n.is_real else False
    è_intero = bool(n.is_integer)
    è_razionale = bool(n.is_rational)
    è_reale = bool(n.is_real)
    print(f"{n}: naturale={è_naturale}, intero={è_intero}, razionale={è_razionale}, reale={è_reale}")

# Verifica esplicita: sqrt(2) è irrazionale?
print("\nsqrt(2) è razionale?", sp.sqrt(2).is_rational)  # None o False, SymPy sa che è irrazionale
print("pi è razionale?", sp.pi.is_rational)
````

## Esercizio svolto e commentato

**Problema:** classifica il numero $x = \dfrac{\sqrt{4}}{2} + \dfrac{3}{2} \times 2$ nell'insieme numerico più piccolo possibile tra $\mathbb{N}, \mathbb{Z}, \mathbb{Q}, \mathbb{R}$, spiegando ogni passaggio.

**Ragionamento:**

Semplifichiamo l'espressione passo per passo, seguendo l'ordine delle operazioni (notebook 0.7).

**Passo 1:** $\sqrt{4} = 2$ (perché $2^2=4$, e per definizione la radice quadrata restituisce il valore non negativo, notebook 0.6).

**Passo 2:** $\dfrac{\sqrt{4}}{2} = \dfrac{2}{2} = 1$.

**Passo 3:** $\dfrac{3}{2}\times2 = 3$ (la moltiplicazione "annulla" la divisione per 2, proprietà delle frazioni notebook 0.3).

**Passo 4:** sommiamo: $1+3=4$.

Quindi $x=4$. Ora classifichiamolo: è un numero **naturale** ($4\in\mathbb{N}$), che è il più piccolo tra gli insiemi della nostra catena $\mathbb{N}\subset\mathbb{Z}\subset\mathbb{Q}\subset\mathbb{R}$. Essendo naturale, appartiene automaticamente anche a tutti gli insiemi successivi (per la catena di inclusioni), ma la risposta più informativa è la **più piccola** in cui si trova.

**Perché cercare "l'insieme più piccolo possibile" è più informativo che dire semplicemente "$x\in\mathbb{R}$":** dire che $x\in\mathbb{R}$ è vero ma dice pochissimo (ci sono infiniti numeri reali "strani"). Dire che $x\in\mathbb{N}$ comunica molta più informazione sulla vera natura del numero — un'espressione con radici e frazioni può, dopo la semplificazione, "nascondere" un numero molto più semplice di quanto sembri all'apparenza.

````python
import sympy as sp

x = sp.sqrt(4)/2 + sp.Rational(3, 2) * 2
x_semplificato = sp.simplify(x)
print("x semplificato:", x_semplificato)
print("è intero?", x_semplificato.is_integer)
print("è >= 0?", x_semplificato >= 0)
````

## Riepilogo finale

- $\mathbb{N}\subset\mathbb{Z}\subset\mathbb{Q}\subset\mathbb{R}$: ogni insieme risolve un limite operativo del precedente (sottrazione, divisione, "buchi" sulla retta).
- Le operazioni di addizione e moltiplicazione sono sempre possibili in tutti e quattro; sottrazione richiede almeno $\mathbb{Z}$; divisione richiede almeno $\mathbb{Q}$.
- $\mathbb{Q}$ è denso in $\mathbb{R}$ ma non completo; $\mathbb{R}$ è sia denso (in sé) sia completo — sono due proprietà distinte, da non confondere.
- Prima di classificare un numero, conviene sempre semplificare l'espressione: la forma "grezza" può nascondere un numero appartenente a un insieme molto più piccolo di quanto sembri.
- Il percorso $\mathbb{N}\to\mathbb{Z}\to\mathbb{Q}\to\mathbb{R}$ non è casuale: ogni allargamento risponde a un bisogno preciso, un modello di come la matematica cresce per rispondere a domande senza risposta.
