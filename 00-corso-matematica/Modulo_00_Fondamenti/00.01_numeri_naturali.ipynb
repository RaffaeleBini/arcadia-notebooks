## Perché ci interessa

Tutto quello che faremo in matematica, dai polinomi agli integrali, si costruisce a partire da qualcosa di molto semplice: la capacità di **contare**.

Quando dici "ho 3 mele" o "in questa stanza ci sono 12 sedie", stai usando i numeri naturali. Sono lo strumento più antico e più intuitivo della matematica, ma proprio per questo vale la pena guardarli con attenzione: capire bene come funzionano i numeri naturali ci darà il linguaggio e le abitudini di pensiero che serviranno per costruire, un passo alla volta, tutti gli altri insiemi numerici (interi, razionali, reali...).

In questo notebook fissiamo le basi: cosa sono i numeri naturali, quali operazioni si possono fare con loro, e quali no (e perché questo "no" sarà proprio il motivo per cui, nel prossimo notebook, introdurremo i numeri interi).

## Teoria, passo dopo passo

### Un esempio concreto

Immagina di dover contare le pecore in un recinto: 0 pecore (recinto vuoto), 1 pecora, 2 pecore, 3 pecore... Il procedimento è sempre lo stesso: parti da un numero e vai al **successivo**, aggiungendo un'unità.

Questo semplice gesto — "andare al numero dopo" — è l'idea centrale dietro ai numeri naturali.

### L'insieme dei numeri naturali

Chiamiamo **numeri naturali** l'insieme

$$\mathbb{N} = \{0, 1, 2, 3, 4, 5, \dots\}$$

usato per contare oggetti. In questo corso includiamo lo 0 tra i naturali (è una convenzione: alcuni libri iniziano da 1, non è un errore, è solo una scelta di notazione — noi seguiamo la convenzione usata in gran parte della matematica moderna e in informatica).

L'idea di "successivo" si può formalizzare così: ogni numero naturale $n$ ha un unico successore, che indichiamo $n+1$. Partendo da 0 e applicando "successivo" ripetutamente si ottengono tutti i naturali:

$$0,\ 0+1=1,\ 1+1=2,\ 2+1=3,\ \dots$$

Questa idea — costruire l'insieme partendo da un elemento base (0) e da un'operazione che genera il "prossimo" elemento — è alla base degli **assiomi di Peano**, la definizione rigorosa di $\mathbb{N}$. Non ci serve il dettaglio completo ora: ci basta portare a casa l'idea che $\mathbb{N}$ è un insieme *costruito passo dopo passo*, e che questa costruzione è ciò che rende possibile un procedimento fondamentale che useremo moltissimo più avanti: il **principio di induzione** (lo vedremo con calma nel notebook 0.20).

### Le operazioni: cosa funziona e cosa no

Su $\mathbb{N}$ possiamo sempre fare due operazioni senza problemi:

- **Addizione**: sommando due naturali si ottiene sempre un naturale. Es. $3 + 5 = 8$.
- **Moltiplicazione**: moltiplicando due naturali si ottiene sempre un naturale. Es. $3 \times 5 = 15$.

Quando un'operazione, applicata a elementi di un insieme, dà sempre come risultato un elemento dello stesso insieme, diciamo che l'insieme è **chiuso** rispetto a quell'operazione. $\mathbb{N}$ è chiuso rispetto ad addizione e moltiplicazione.

Ma la **sottrazione** è un problema:

$$5 - 3 = 2 \quad \text{(ok, è un naturale)}$$
$$3 - 5 = ? \quad \text{(non esiste nessun naturale che dia questo risultato)}$$

$\mathbb{N}$ **non è chiuso** rispetto alla sottrazione. Questa è un'osservazione semplice ma importantissima: è esattamente il motivo per cui, nel prossimo notebook, allargheremo l'insieme dei numeri introducendo i **numeri interi** $\mathbb{Z}$, che includono anche i negativi e risolvono questo problema.

### L'ordine e il buon ordinamento

I naturali sono **ordinati**: dati due naturali diversi, uno è sempre minore dell'altro ($2 < 5$, per esempio). Questo ci sembra ovvio, ma ha una conseguenza speciale, chiamata **principio del buon ordinamento**:

> Ogni sottoinsieme non vuoto di $\mathbb{N}$ ha un elemento minimo.

Per esempio, l'insieme dei naturali pari maggiori di 10 è $\{12, 14, 16, \dots\}$: il suo minimo è 12. Questa proprietà sembra banale ma **non vale** per altri insiemi numerici che vedremo presto (per esempio $\mathbb{Q}$, i razionali, non ha questa proprietà). È uno dei tanti motivi per cui $\mathbb{N}$ ha un ruolo speciale: è il "mattone" da cui si parte.

````python
# Verifichiamo con Python alcune delle proprietà viste sopra.
# Useremo semplici test su numeri naturali per "vedere" la teoria in azione.

# 1. Chiusura di addizione e moltiplicazione: proviamo con diverse coppie
coppie = [(3, 5), (0, 7), (12, 4)]
for a, b in coppie:
    somma = a + b
    prodotto = a * b
    # somma e prodotto di due naturali sono sempre naturali (>= 0, interi)
    print(f"{a} + {b} = {somma}  (naturale: {somma >= 0})")
    print(f"{a} * {b} = {prodotto}  (naturale: {prodotto >= 0})")

# 2. La sottrazione invece può "uscire" da N
a, b = 3, 5
differenza = a - b
print(f"\n{a} - {b} = {differenza}  -> non è un numero naturale, è negativo!")

# 3. Buon ordinamento: troviamo il minimo di un sottoinsieme di N
sottoinsieme = {n for n in range(0, 50) if n % 2 == 0 and n > 10}  # pari maggiori di 10
print(f"\nSottoinsieme: {sorted(sottoinsieme)[:5]}...")
print(f"Elemento minimo: {min(sottoinsieme)}")
````

## Esercizio svolto e commentato

**Problema:** dimostra che la somma di due numeri naturali pari è sempre un numero naturale pari.

**Ragionamento:**

Un numero naturale $n$ è pari se possiamo scriverlo come $n = 2k$, per qualche naturale $k$ (cioè è "due volte qualcosa"). Questa è la definizione di "pari": non ci basiamo su esempi, vogliamo un argomento che funzioni per **tutti** i numeri pari, non solo per quelli che proviamo a mano.

Prendiamo allora due numeri pari qualsiasi. Chiamiamoli $a$ e $b$. Per definizione possiamo scrivere:

$$a = 2k_1, \qquad b = 2k_2$$

dove $k_1$ e $k_2$ sono naturali (magari diversi tra loro: stiamo prendendo due numeri pari generici, non necessariamente uguali).

Sommiamo:

$$a + b = 2k_1 + 2k_2$$

Qui usiamo la proprietà distributiva al contrario (raccogliamo il fattore comune 2):

$$a + b = 2(k_1 + k_2)$$

Ora, $k_1 + k_2$ è a sua volta un numero naturale (perché $\mathbb{N}$ è chiuso rispetto all'addizione, come abbiamo visto sopra!). Chiamiamolo $k_3 = k_1 + k_2$. Allora:

$$a + b = 2 k_3$$

cioè $a+b$ è "due volte un naturale": è pari, per definizione.

**Perché questo ragionamento è convincente e un esempio numerico no:** se avessimo solo verificato $4 + 6 = 10$ (pari), avremmo dimostrato la proprietà solo per *quella* coppia di numeri. Scrivendo $a$ e $b$ in generale come $2k_1$ e $2k_2$, l'argomento vale per **ogni** possibile coppia di numeri pari, perché non abbiamo usato nessuna caratteristica specifica di 4 e 6.

````python
# Verifichiamo il risultato su alcuni esempi (la verifica numerica non è una dimostrazione,
# ma è un buon modo per controllare che il ragionamento sia corretto)

def è_pari(n):
    return n % 2 == 0

test = [(4, 6), (0, 8), (10, 100), (2, 2)]
for a, b in test:
    assert è_pari(a) and è_pari(b), "stiamo testando solo su numeri pari"
    somma = a + b
    print(f"{a} (pari) + {b} (pari) = {somma}  -> pari: {è_pari(somma)}")
````

**1.** Un numero dispari si scrive $2k+1$. Prendiamo due dispari $a = 2k_1+1$ e $b = 2k_2+1$. Sommando:
$$a+b = 2k_1+1+2k_2+1 = 2k_1+2k_2+2 = 2(k_1+k_2+1)$$
Poiché $k_1+k_2+1$ è un naturale, $a+b$ è "due volte un naturale", quindi pari.

**2.** Se $a=2k_1$ e $b=2k_2$, allora $a \cdot b = 4 k_1 k_2 = 2(2 k_1 k_2)$, che è "due volte un naturale" ($2k_1k_2$ è naturale), quindi pari.

**3.** Cerchiamo il più piccolo naturale il cui quadrato superi 50. Proviamo: $7^2=49$ (non basta, $49 \le 50$), $8^2=64$ (basta, $64>50$). Quindi il minimo è $8$.

**4.** Il buon ordinamento dice solo che esiste sempre un **minimo**, non dice nulla su un massimo. L'insieme dei pari $\{0,2,4,6,\dots\}$ non ha massimo perché per ogni pari $n$ esiste sempre un pari più grande ($n+2$): l'insieme è infinito e "sale" senza fermarsi mai. Il buon ordinamento riguarda solo l'esistenza del minimo, proprietà che vale sempre in $\mathbb{N}$ indipendentemente dal fatto che l'insieme sia finito o infinito.

## Riepilogo finale

- I numeri naturali $\mathbb{N} = \{0, 1, 2, 3, \dots\}$ nascono dall'idea di contare e di "andare al numero successivo".
- $\mathbb{N}$ è **chiuso** rispetto ad addizione e moltiplicazione: il risultato è sempre un naturale.
- $\mathbb{N}$ **non è chiuso** rispetto alla sottrazione: $3 - 5$ non ha soluzione in $\mathbb{N}$. Questo limite motiva l'introduzione dei numeri interi.
- $\mathbb{N}$ gode del **principio del buon ordinamento**: ogni suo sottoinsieme non vuoto ha un elemento minimo. Questa proprietà sarà alla base dell'induzione matematica.
- Dimostrare una proprietà "in generale" (con lettere, come $a=2k$) è molto più forte che verificarla su singoli esempi numerici.
