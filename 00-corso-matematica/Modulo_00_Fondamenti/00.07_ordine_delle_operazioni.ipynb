# 0.7 — Ordine delle operazioni

**Prerequisiti:** 0.6 Frazioni, potenze e radici.


## Perché ci interessa

Considera l'espressione $2 + 3 \times 4$. Se la leggi da sinistra a destra senza regole, potresti calcolare $2+3=5$ e poi $5\times4=20$. Ma se prima fai $3\times4=12$ e poi $2+12=14$, ottieni un risultato diverso.

Un'espressione matematica deve avere **un solo risultato possibile**, altrimenti la matematica non sarebbe uno strumento affidabile per comunicare calcoli. Per questo esiste una convenzione universale, condivisa da tutti, su **in che ordine eseguire le operazioni**. Non è una regola dettata dal caso: vedremo che nasce dal voler rendere le espressioni compatte senza perdere significato.

## Teoria, passo dopo passo

### Perché il prodotto "viene prima" della somma

Pensiamo a cosa significa davvero $2 + 3 \times 4$. Il simbolo $3\times4$ è un modo compatto per scrivere "$3$ ripetuto $4$ volte", cioè $3+3+3+3$. Quindi:

$$2 + 3 \times 4 = 2 + (3+3+3+3) = 2 + 12 = 14$$

La moltiplicazione **è già**, nella sua definizione, un blocco unico che si "consuma" prima di essere sommato ad altro. Non è una regola arbitraria imposta dall'esterno: riflette il significato stesso dell'operazione. Lo stesso vale per le potenze rispetto alla moltiplicazione: $2 \times 3^2 = 2 \times (3\times3) = 2\times9=18$, non $(2\times3)^2=36$.

### La gerarchia standard

L'ordine convenzionale, dal livello più "stretto" (si esegue per primo) al più "largo":

1. **Parentesi** — quello che è dentro si calcola per primo, sempre, perché le parentesi servono proprio a "forzare" un ordine diverso da quello standard.
2. **Potenze e radici**
3. **Moltiplicazioni e divisioni** (stesso livello: si eseguono da sinistra a destra nell'ordine in cui compaiono)
4. **Addizioni e sottrazioni** (stesso livello: da sinistra a destra)

Un acronimo comune per ricordarla (in inglese) è PEMDAS (Parentheses, Exponents, Multiplication/Division, Addition/Subtraction) — non serve impararlo a memoria, basta ricordare **perché** questo ordine ha senso, come abbiamo visto sopra.

### Un'insidia comune: divisione e moltiplicazione hanno la stessa priorità

Molti studenti pensano che la moltiplicazione "venga prima" della divisione. Non è così: hanno la **stessa priorità**, quindi si eseguono nell'ordine in cui compaiono, da sinistra a destra.

$$20 \div 4 \times 5 = (20 \div 4) \times 5 = 5 \times 5 = 25$$

Se invece si volesse dividere per $4\times5$, servirebbero le parentesi: $20 \div (4\times5) = 20\div20=1$. Un risultato completamente diverso! Questo mostra quanto sia importante scrivere le espressioni senza ambiguità.

### Le parentesi come "cambio di priorità"

Quando vogliamo che un'addizione venga calcolata *prima* di una moltiplicazione (contro l'ordine standard), usiamo le parentesi:

$$(2+3)\times4 = 5\times4=20 \qquad \ne \qquad 2+3\times4=14$$

Le parentesi non "creano" una nuova regola: semplicemente dicono "qui dentro, calcola tutto prima di procedere".

````python
import sympy as sp

# Verifichiamo con SymPy (che rispetta rigorosamente l'ordine delle operazioni)
espressione1 = 2 + 3 * 4
print(f"2 + 3*4 = {espressione1}")  # atteso 14, non 20

espressione2 = (2 + 3) * 4
print(f"(2+3)*4 = {espressione2}")  # atteso 20

espressione3 = 2 * 3**2
print(f"2 * 3**2 = {espressione3}")  # atteso 18, non 36

# Divisione e moltiplicazione: stessa priorità, da sinistra a destra
espressione4 = 20 / 4 * 5
print(f"20 / 4 * 5 = {espressione4}")  # atteso 25.0

espressione5 = 20 / (4 * 5)
print(f"20 / (4*5) = {espressione5}")  # atteso 1.0, molto diverso dal precedente!
````

## Esercizio svolto e commentato

**Problema:** calcola $\dfrac{2 + 3\times(4-1)^2}{5}$, mostrando ogni passaggio nell'ordine corretto.

**Ragionamento:**

Seguiamo rigorosamente la gerarchia: parentesi, poi potenze, poi moltiplicazioni/divisioni, poi addizioni/sottrazioni. La frazione stessa si comporta come una "grande parentesi": numeratore e denominatore vanno calcolati per intero prima di dividere.

**Passo 1 — parentesi più interna:** $4-1=3$. L'espressione diventa:

$$\frac{2+3\times3^2}{5}$$

**Passo 2 — potenza:** $3^2=9$. Diventa:

$$\frac{2+3\times9}{5}$$

**Passo 3 — moltiplicazione:** $3\times9=27$. Diventa:

$$\frac{2+27}{5}$$

**Passo 4 — addizione (dentro il numeratore, che va completato prima di dividere):** $2+27=29$. Diventa:

$$\frac{29}{5}$$

**Passo 5 — divisione finale:** $\frac{29}{5}=5.8$

**Perché seguiamo esattamente questo ordine e non un altro:** ogni passaggio "consuma" l'operazione di priorità più alta rimasta. Se avessimo, per esempio, sommato $2+3$ prima della potenza, avremmo violato la gerarchia e ottenuto un risultato sbagliato. La disciplina nel seguire l'ordine, passo per passo, è ciò che garantisce di arrivare sempre allo stesso risultato corretto.

````python
import sympy as sp

risultato = sp.Rational(2 + 3 * (4 - 1)**2, 5)
print(f"Risultato: {risultato} = {float(risultato)}")  # atteso 29/5 = 5.8
````

## Riepilogo finale

- L'ordine delle operazioni non è arbitrario: riflette il significato stesso di ogni operazione (es. la moltiplicazione è "somma ripetuta", quindi va calcolata come blocco unico).
- Gerarchia dal più forte al più debole: parentesi → potenze/radici → moltiplicazioni/divisioni → addizioni/sottrazioni.
- Moltiplicazione e divisione hanno la **stessa priorità**: si eseguono da sinistra a destra nell'ordine in cui compaiono (lo stesso vale per addizione e sottrazione).
- Le parentesi servono a "forzare" un ordine diverso da quello standard.
- Una frazione con espressioni complesse a numeratore/denominatore si comporta come se ci fossero parentesi implicite attorno a numeratore e denominatore.
