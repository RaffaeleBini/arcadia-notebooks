## Perché ci interessa

La distanza Terra-Sole è circa 150.000.000.000 metri. La massa di un elettrone è circa 0.00000000000000000000000000000091 kg. Scrivere e leggere numeri così, con tutti quegli zeri, è scomodo e pieno di rischi di errore (basta contare male uno zero per sbagliare di un fattore 10).

Ci serve un modo compatto per scrivere numeri molto grandi o molto piccoli, che sfrutti le potenze di 10 (notebook 0.6) per "nascondere" tutti quegli zeri in un esponente.

## Teoria, passo dopo passo

### L'idea di base: le potenze di 10

Ricordiamo come funzionano le potenze di 10:

$$10^3 = 1000 \qquad 10^{-3} = \frac{1}{1000}=0.001$$

Moltiplicare per $10^n$ (con $n$ positivo) sposta la virgola di $n$ posizioni verso destra; moltiplicare per $10^{-n}$ la sposta di $n$ posizioni verso sinistra. Questa osservazione è la chiave della notazione scientifica.

### La forma della notazione scientifica

Un numero in **notazione scientifica** si scrive come:

$$m \times 10^n$$

dove $m$ (la "mantissa") è un numero con $1 \le |m| < 10$ (cioè una sola cifra non nulla prima della virgola), e $n$ è un intero (l'esponente).

Esempio: $150.000.000.000 = 1.5 \times 10^{11}$. Verifichiamo: $1.5 \times 10^{11} = 1.5 \times 100.000.000.000 = 150.000.000.000$. ✓

Per numeri molto piccoli, l'esponente sarà negativo: $0.00000000000000000000000000000091 = 9.1 \times 10^{-31}$ (la massa dell'elettrone in kg).

### Come convertire un numero in notazione scientifica

Il procedimento è sempre lo stesso: sposta la virgola fino ad avere una sola cifra significativa prima di essa, e conta quante posizioni hai spostato — quel conteggio è l'esponente.

- Se sposti la virgola **verso sinistra** (numero grande, come 150.000.000.000), l'esponente è **positivo** (perché per tornare al numero originale dovrai moltiplicare per una potenza di 10 che sposta la virgola verso destra).
- Se sposti la virgola **verso destra** (numero piccolo, come 0.00000000000000000000000000000091), l'esponente è **negativo**.

Esempio passo passo con $0.0042$: spostiamo la virgola verso destra fino a ottenere $4.2$ (una sola cifra prima della virgola). Abbiamo spostato la virgola di 3 posizioni verso destra, quindi:

$$0.0042 = 4.2 \times 10^{-3}$$

### Operazioni in notazione scientifica

Moltiplicare e dividere numeri in notazione scientifica è comodo, perché si usano le proprietà delle potenze (notebook 0.6):

$$(a \times 10^m) \times (b \times 10^n) = (a\times b) \times 10^{m+n}$$

Esempio: $(2 \times 10^5) \times (3 \times 10^3) = 6 \times 10^8$.

Attenzione: se il prodotto delle mantisse esce dal range $[1,10)$, bisogna "riaggiustare". Esempio: $(4\times10^5)\times(5\times10^3) = 20\times10^8$, ma $20$ non è una mantissa valida (deve essere $<10$): riscriviamo $20=2\times10^1$, quindi il risultato finale è $2\times10^{1}\times10^8=2\times10^9$.

````python
import sympy as sp

# SymPy può mostrare i numeri in notazione scientifica con sp.N e la formattazione Python
distanza_terra_sole = 150_000_000_000
print(f"{distanza_terra_sole:.2e}")  # notazione scientifica con 2 cifre decimali nella mantissa

massa_elettrone = 0.00000000000000000000000000000091
print(f"{massa_elettrone:.2e}")

# Conversione manuale, verificata con sympy
mantissa, esponente = sp.Rational('4.2'), -3
valore_ricostruito = mantissa * 10**esponente
print(f"\n4.2 * 10^-3 = {float(valore_ricostruito)}  (atteso 0.0042)")

# Moltiplicazione in notazione scientifica, con "riaggiustamento" della mantissa
a, m = 4, 5   # 4 * 10^5
b, n = 5, 3   # 5 * 10^3
prodotto_mantisse = a * b
esponente_somma = m + n
print(f"\n({a}e{m}) * ({b}e{n}) = {prodotto_mantisse}e{esponente_somma} -> da riaggiustare")
# 20 non è una mantissa valida (deve essere tra 1 e 10): riscriviamo 20 = 2 * 10^1
mantissa_corretta = prodotto_mantisse / 10
esponente_corretto = esponente_somma + 1
print(f"Forma corretta: {mantissa_corretta}e{esponente_corretto}")
````

## Esercizio svolto e commentato

**Problema:** la popolazione mondiale è circa $8 \times 10^9$ persone. Se ognuna bevesse in media $2$ litri d'acqua al giorno, quanti litri d'acqua si consumerebbero in un giorno, in tutto il mondo? Esprimi il risultato in notazione scientifica.

**Ragionamento:**

Vogliamo calcolare (numero di persone) × (litri per persona):

$$(8 \times 10^9) \times 2$$

Il numero $2$ non è scritto in notazione scientifica, ma possiamo pensarlo come $2\times10^0$ (perché $10^0=1$, non cambia nulla). Moltiplichiamo mantisse tra loro e sommiamo gli esponenti:

$$(8\times2) \times 10^{9+0} = 16 \times 10^9$$

Il numero $16$ non è una mantissa valida (deve essere $1\le m <10$). Riscriviamo $16 = 1.6\times10^1$:

$$16\times10^9 = 1.6\times10^1\times10^9 = 1.6\times10^{10}$$

Il consumo giornaliero sarebbe circa $1.6\times10^{10}$ litri, cioè 16 miliardi di litri.

**Perché "riaggiustare" la mantissa è importante e non solo una formalità:** la notazione scientifica è utile proprio perché permette di confrontare a colpo d'occhio l'ordine di grandezza di due numeri guardando solo l'esponente. Se lasciassimo mantisse "sballate" come $16\times10^9$, perderemmo questa capacità di confronto immediato (potremmo confondere $16\times10^9$ con qualcosa dell'ordine di $10^9$, mentre in realtà è dell'ordine di $10^{10}$).

````python
import sympy as sp

popolazione = sp.Rational(8) * 10**9
consumo_procapite = 2
consumo_totale = popolazione * consumo_procapite
print(f"Consumo totale: {float(consumo_totale):.2e} litri")
````

## Riepilogo finale

- La notazione scientifica scrive un numero come $m \times 10^n$, con $1\le|m|<10$: comprime numeri molto grandi o molto piccoli in una forma leggibile.
- Per convertire, si sposta la virgola fino a lasciare una sola cifra significativa prima di essa: il numero di posizioni spostate è l'esponente (positivo se il numero era grande, negativo se era piccolo).
- Moltiplicare/dividere in notazione scientifica è comodo: si opera sulle mantisse e si sommano/sottraggono gli esponenti (proprietà delle potenze, notebook 0.6).
- Se la mantissa del risultato esce dal range $[1,10)$, va "riaggiustata" spostando ancora la virgola e correggendo l'esponente di conseguenza.
