## Perché ci interessa

Finora abbiamo parlato di insiemi "isolati". Ma spesso la matematica descrive **legami** tra elementi di insiemi diversi (o dello stesso insieme): "$x$ è padre di $y$", "$x$ è minore di $y$", "$x$ è il quadrato di $y$". Questi legami si chiamano **relazioni**.

Un tipo particolare di relazione — quella in cui ogni elemento di partenza è collegato a **uno e un solo** elemento di arrivo — si chiama **funzione**, ed è probabilmente il concetto più importante di tutta la matematica che studieremo da qui in avanti (le dedicheremo l'intero Modulo 3). In questo notebook poniamo le basi, partendo dal caso generale delle relazioni.

## Teoria, passo dopo passo

### Coppie ordinate e prodotto cartesiano

Prima di parlare di relazioni, ci serve un modo per "accoppiare" elementi. Una **coppia ordinata** $(a,b)$ è diversa da $(b,a)$ se $a\ne b$ (a differenza degli insiemi, dove l'ordine non conta — notebook 0.12): l'ordine qui è essenziale.

Dati due insiemi $A$ e $B$, il **prodotto cartesiano** $A \times B$ è l'insieme di tutte le coppie ordinate con primo elemento in $A$ e secondo elemento in $B$:

$$A \times B = \{(a,b) : a \in A,\ b \in B\}$$

Esempio: se $A=\{1,2\}$ e $B=\{x,y\}$, allora $A\times B = \{(1,x),(1,y),(2,x),(2,y)\}$: 4 coppie (in generale, se $A$ ha $m$ elementi e $B$ ne ha $n$, $A\times B$ ne ha $m\times n$ — da cui il nome "prodotto").

### Cos'è una relazione

Una **relazione** $R$ da $A$ a $B$ è semplicemente un sottoinsieme del prodotto cartesiano: $R \subseteq A \times B$. Se $(a,b) \in R$, diciamo che "$a$ è in relazione con $b$", scritto anche $a\,R\,b$.

Esempio concreto: $A=B=\mathbb{N}$, e $R$="essere minore di" è l'insieme $R = \{(a,b) \in \mathbb{N}\times\mathbb{N} : a<b\}$. La coppia $(2,5)$ appartiene a $R$ (perché $2<5$), ma $(5,2)$ no.

### Le funzioni come caso speciale delle relazioni

Una **funzione** $f$ da $A$ a $B$ (scritta $f: A \to B$) è una relazione con una proprietà speciale, molto restrittiva:

$$\forall a \in A,\ \exists! b \in B \ \text{tale che}\ (a,b) \in f$$

("per ogni $a$ in $A$, esiste esattamente un $b$ in $B$ collegato ad $a$" — ricordi il simbolo $\exists!$, "esiste un unico", visto nel notebook 0.17?). In altre parole: **ogni elemento di $A$ deve avere esattamente un'immagine**, né zero né più di una. Scriviamo $b=f(a)$ per indicare l'unico elemento associato ad $a$.

Perché questa proprietà è così importante? Perché è ciò che rende una funzione **prevedibile e calcolabile**: dato $a$, sai sempre esattamente cosa aspettarti, senza ambiguità. Approfondiremo tutto questo (dominio, codominio, iniettività...) con calma nel Modulo 3: qui ci basta l'intuizione di base.

### Un esempio di relazione che NON è una funzione, e uno che lo è

- $R = \{(x,y) \in \mathbb{R}\times\mathbb{R} : x^2+y^2=1\}$ (punti su una circonferenza di raggio 1) **non è una funzione**: per esempio $x=0$ è collegato sia a $y=1$ sia a $y=-1$ (due elementi $y$ diversi per lo stesso $x$), violando l'unicità richiesta.
- $f = \{(x,y) \in \mathbb{R}\times\mathbb{R} : y = x^2\}$ **è una funzione**: per ogni $x$ c'è un solo $y$ possibile, $y=x^2$ (elevare al quadrato dà sempre un risultato unico).

### Rappresentazione grafica: il test della retta verticale

Se disegniamo una relazione su un piano cartesiano (che studieremo in dettaglio nel notebook 2.1), c'è un modo visivo rapido per riconoscere se è una funzione: tracciamo una retta verticale in un punto qualsiasi. Se questa retta incontra il grafico della relazione **in più di un punto**, allora la relazione non è una funzione (perché quella $x$ avrebbe più di un $y$ associato). Se ogni retta verticale incontra il grafico in **al massimo un punto**, allora è una funzione.

````python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Relazione NON funzione: circonferenza x^2+y^2=1
theta = np.linspace(0, 2*np.pi, 200)
x_cerchio = np.cos(theta)
y_cerchio = np.sin(theta)
axes[0].plot(x_cerchio, y_cerchio, color='tab:red')
axes[0].axvline(x=0, color='gray', linestyle='--')  # retta verticale di test
axes[0].plot([0, 0], [1, -1], 'ko')  # evidenzia i DUE punti di intersezione
axes[0].set_title("x² + y² = 1 : NON è una funzione\n(la retta verticale tocca 2 punti)")
axes[0].set_aspect('equal')
axes[0].grid(True, alpha=0.3)

# Relazione che È una funzione: y = x^2
x_parabola = np.linspace(-2, 2, 200)
y_parabola = x_parabola**2
axes[1].plot(x_parabola, y_parabola, color='tab:blue')
axes[1].axvline(x=1, color='gray', linestyle='--')  # retta verticale di test
axes[1].plot([1], [1], 'ko')  # un SOLO punto di intersezione
axes[1].set_title("y = x² : è una funzione\n(ogni retta verticale tocca 1 punto)")
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Esercizio svolto e commentato

**Problema:** dati $A=\{1,2,3\}$ e $B=\{a,b\}$, elenca tutti gli elementi di $A \times B$, e poi stabilisci se la relazione $R = \{(1,a),(2,a),(3,b)\}$ è una funzione da $A$ a $B$.

**Ragionamento:**

**Prodotto cartesiano:** per definizione, $A\times B$ contiene tutte le coppie con primo elemento in $A$ e secondo in $B$. Con $A$ che ha 3 elementi e $B$ che ne ha 2, ci aspettiamo $3\times2=6$ coppie:

$$A\times B = \{(1,a),(1,b),(2,a),(2,b),(3,a),(3,b)\}$$

**Verifica se $R$ è una funzione:** la definizione richiede che **ogni** elemento di $A$ compaia **esattamente una volta** come primo elemento di una coppia in $R$. Controlliamo elemento per elemento:

- $1$: compare in $(1,a) \in R$. Una sola volta. ✓
- $2$: compare in $(2,a) \in R$. Una sola volta. ✓
- $3$: compare in $(3,b) \in R$. Una sola volta. ✓

Ogni elemento di $A$ ha esattamente un'immagine in $R$: **sì, è una funzione**. Possiamo scriverla come $f(1)=a$, $f(2)=a$, $f(3)=b$ (nota che $a$ è immagine sia di $1$ sia di $2$: **questo è permesso**, la definizione di funzione richiede un solo $b$ per ogni $a$, ma non vieta che lo stesso $b$ sia raggiunto da più $a$ diversi).

**Perché controlliamo "ogni elemento di $A$" e non "ogni elemento di $B$":** la definizione di funzione riguarda solo la direzione da $A$ verso $B$: ogni punto di partenza deve avere esattamente un punto di arrivo. Non c'è invece nessun vincolo su quante volte un elemento di $B$ venga "raggiunto" (può essere raggiunto da più elementi di $A$, o anche da nessuno).

````python
A = {1, 2, 3}
B = {'a', 'b'}

prodotto_cartesiano = {(x, y) for x in A for y in B}
print("A x B =", prodotto_cartesiano)
print("Numero di coppie:", len(prodotto_cartesiano), "(atteso 3*2=6)")

R = {(1, 'a'), (2, 'a'), (3, 'b')}

def è_funzione(relazione, dominio):
    # controlla che ogni elemento del dominio compaia come primo elemento
    # in esattamente una coppia della relazione
    for elemento in dominio:
        immagini = [coppia[1] for coppia in relazione if coppia[0] == elemento]
        if len(immagini) != 1:
            return False, elemento
    return True, None

risultato, problema = è_funzione(R, A)
print("\nR è una funzione da A a B?", risultato)
````

## Riepilogo finale

- Il **prodotto cartesiano** $A\times B$ è l'insieme di tutte le coppie ordinate $(a,b)$ con $a\in A$, $b\in B$.
- Una **relazione** da $A$ a $B$ è un qualunque sottoinsieme di $A\times B$: descrive un legame generico tra elementi.
- Una **funzione** è una relazione speciale: ogni elemento di $A$ deve avere **esattamente un'immagine** in $B$ (né zero, né più di una).
- Il test della retta verticale è un modo visivo rapido per riconoscere se un grafico rappresenta una funzione.
- Una funzione può assegnare la stessa immagine a elementi diversi del dominio: il vincolo di unicità vale solo nella direzione dominio→immagine, non viceversa.
