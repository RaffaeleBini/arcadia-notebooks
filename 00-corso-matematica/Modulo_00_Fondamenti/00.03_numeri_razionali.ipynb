# 0.3 — Numeri razionali

**Prerequisiti:** 0.1 Numeri naturali, 0.2 Numeri interi.


## Perché ci interessa

Abbiamo visto che $\mathbb{Z}$ risolve il problema della sottrazione, ma non quello della divisione: $7 \div 2$ non è un numero intero. Eppure ha senso dire "dividiamo 7 pizze tra 2 persone": ognuno prende $3$ pizze e mezza.

Ci serve un insieme di numeri che rappresenti anche le **parti di un intero**: le frazioni. Questo insieme si chiama insieme dei **numeri razionali**.

## Teoria, passo dopo passo

### Dalle divisioni alle frazioni

Quando dividiamo $a$ oggetti in $b$ parti uguali (con $b \ne 0$), il risultato si scrive come frazione:

$$\frac{a}{b}$$

Un numero razionale è, per definizione, un numero che si può scrivere come rapporto tra due interi $a$ e $b$, con $b \ne 0$:

$$\mathbb{Q} = \left\{ \frac{a}{b} : a, b \in \mathbb{Z},\ b \ne 0 \right\}$$

(la lettera $\mathbb{Q}$ viene da *quoziente*). Notiamo che $\mathbb{Z} \subset \mathbb{Q}$: ogni intero $n$ è anche razionale, perché si scrive come $\frac{n}{1}$.

### La stessa quantità, tante scritture diverse

Una cosa fondamentale delle frazioni è che **rappresentazioni diverse possono indicare lo stesso numero**: $\frac{1}{2}$, $\frac{2}{4}$ e $\frac{50}{100}$ indicano tutte "una metà". Diciamo che due frazioni $\frac{a}{b}$ e $\frac{c}{d}$ sono **equivalenti** se:

$$a \times d = b \times c$$

Per esempio $\frac{1}{2}$ e $\frac{2}{4}$: $1 \times 4 = 4$ e $2 \times 2 = 4$, coincidono, quindi sono equivalenti. Questa regola viene dal fatto che moltiplicare numeratore e denominatore per lo stesso numero (diverso da 0) non cambia il valore della frazione: $\frac{a}{b} = \frac{a \cdot k}{b \cdot k}$.

Tra tutte le scritture equivalenti, quella con numeratore e denominatore senza fattori comuni si chiama **frazione ridotta ai minimi termini** (es. $\frac{1}{2}$ per la famiglia di frazioni equivalenti a $\frac{2}{4}, \frac{3}{6}$, ecc.).

### Le operazioni tra frazioni

- **Moltiplicazione**: si moltiplicano numeratori tra loro e denominatori tra loro: $\frac{a}{b} \times \frac{c}{d} = \frac{a c}{b d}$.
- **Addizione**: bisogna prima portare le frazioni allo stesso denominatore (denominatore comune), poi si sommano i numeratori: $\frac{a}{b} + \frac{c}{d} = \frac{ad + bc}{bd}$.

  Perché funziona così? Perché possiamo sommare solo "quantità dello stesso tipo": $\frac{1}{2}$ (mezzi) e $\frac{1}{3}$ (terzi) non si sommano direttamente, ma se li riscriviamo entrambi in **sesti** ($\frac{3}{6}$ e $\frac{2}{6}$), diventano confrontabili e sommabili: $\frac{3}{6}+\frac{2}{6}=\frac{5}{6}$.

- **Divisione**: dividere per una frazione equivale a moltiplicare per il suo "reciproco" (la frazione capovolta): $\frac{a}{b} \div \frac{c}{d} = \frac{a}{b} \times \frac{d}{c}$, purché $c \ne 0$.

### Cosa manca ancora

Con $\mathbb{Q}$ possiamo finalmente dividere (tranne che per 0, che resta sempre vietato: non esiste nessun numero che, moltiplicato per 0, dia un valore diverso da 0). Sembra che il lavoro sia finito... ma non è così: esistono numeri, come $\sqrt{2}$ o $\pi$, che **non si possono scrivere come frazione di interi**. Questo sarà il punto di partenza del prossimo notebook, sui numeri reali.

````python
from fractions import Fraction

# Python ha un tipo nativo per le frazioni: Fraction, che semplifica automaticamente
a = Fraction(1, 2)
b = Fraction(2, 4)
print(f"{a} è equivalente a {b}? {a == b}")  # Fraction riduce automaticamente ai minimi termini

# Verifica della regola di equivalenza a*d == b*c "a mano"
def equivalenti(frazione1, frazione2):
    a, b = frazione1
    c, d = frazione2
    return a * d == b * c

print("(1,2) equivalente a (50,100)?", equivalenti((1, 2), (50, 100)))

# Operazioni tra frazioni
somma = Fraction(1, 2) + Fraction(1, 3)
print(f"1/2 + 1/3 = {somma}")  # atteso 5/6

divisione = Fraction(3, 4) / Fraction(2, 5)
print(f"(3/4) / (2/5) = {divisione}")  # atteso 15/8
````

## Esercizio svolto e commentato

**Problema:** calcola $\frac{2}{3} + \frac{1}{4} - \frac{1}{6}$, mostrando ogni passaggio.

**Ragionamento:**

Per sommare/sottrarre frazioni serve un denominatore comune. Cerchiamo il **minimo comune multiplo** (m.c.m.) tra $3$, $4$ e $6$, così lavoriamo con i numeri più piccoli possibili (non è obbligatorio usare il minimo, ma rende i calcoli più semplici).

I multipli di questi numeri: il m.c.m. di 3, 4, 6 è $12$ (è il più piccolo numero divisibile per tutti e tre).

Riscriviamo ogni frazione con denominatore 12:

$$\frac{2}{3} = \frac{2 \times 4}{3 \times 4} = \frac{8}{12}, \qquad \frac{1}{4} = \frac{1 \times 3}{4 \times 3} = \frac{3}{12}, \qquad \frac{1}{6} = \frac{1 \times 2}{6 \times 2} = \frac{2}{12}$$

Ora che i denominatori coincidono, possiamo sommare/sottrarre direttamente i numeratori:

$$\frac{8}{12} + \frac{3}{12} - \frac{2}{12} = \frac{8+3-2}{12} = \frac{9}{12}$$

Infine riduciamo ai minimi termini: $9$ e $12$ hanno il fattore comune $3$, quindi:

$$\frac{9}{12} = \frac{9 \div 3}{12 \div 3} = \frac{3}{4}$$

**Perché cerchiamo il m.c.m. e non un denominatore comune qualsiasi:** funzionerebbe anche con un denominatore comune più grande (per esempio $3\times4\times6=72$), ma i numeri da maneggiare sarebbero più grandi e il rischio di errori di calcolo aumenterebbe. Il m.c.m. è la scelta più efficiente, non l'unica possibile.

````python
from fractions import Fraction

risultato = Fraction(2, 3) + Fraction(1, 4) - Fraction(1, 6)
print(f"2/3 + 1/4 - 1/6 = {risultato}")  # atteso 3/4
````

Questa proprietà si chiama **densità** di $\mathbb{Q}$: tra due razionali qualsiasi (per quanto vicini) c'è sempre un altro razionale.

## Riepilogo finale

- I numeri razionali $\mathbb{Q}$ sono i rapporti $\frac{a}{b}$ tra due interi, con $b \ne 0$: nascono per rendere sempre possibile la divisione.
- $\mathbb{Z} \subset \mathbb{Q}$: ogni intero è anche razionale ($n = \frac{n}{1}$).
- Frazioni diverse possono rappresentare lo stesso numero (equivalenza): $\frac{a}{b}=\frac{c}{d} \iff ad=bc$.
- Per sommare frazioni serve un denominatore comune; per dividere si moltiplica per il reciproco.
- $\mathbb{Q}$ è **denso**: tra due razionali distinti ce n'è sempre un altro. Ma, come vedremo nel prossimo notebook, questo non basta a "riempire" completamente la retta dei numeri.
