# 0.8 — Valore assoluto

**Prerequisiti:** 0.2 Numeri interi, 0.4 Numeri reali.


## Perché ci interessa

A volte non ci interessa il segno di un numero, ma solo "quanto è grande" o "quanto dista da zero". Se la temperatura è cambiata di $-5$ gradi o di $+5$ gradi, in entrambi i casi diciamo che è cambiata "di 5 gradi": il segno ci dice la direzione, ma la grandezza del cambiamento è la stessa.

Ci serve uno strumento di notazione che catturi questa idea di "grandezza senza segno": il **valore assoluto**. Lo abbiamo già incontrato di sfuggita nel notebook 0.6 (parlando di $\sqrt{a^2}$): ora lo formalizziamo.

## Teoria, passo dopo passo

### Un esempio concreto: la distanza sulla retta numerica

Immagina la retta dei numeri reali, con lo 0 al centro. Il numero $5$ dista $5$ unità da $0$. Il numero $-5$ dista **anche lui** $5$ unità da $0$, solo dal lato opposto. Il valore assoluto cattura esattamente questa "distanza da zero", ignorando la direzione.

### La definizione

Il valore assoluto di un numero reale $x$, indicato $|x|$, è definito così:

$$|x| = \begin{cases} x & \text{se } x \ge 0 \\ -x & \text{se } x < 0 \end{cases}$$

Leggiamola con attenzione: se $x$ è già positivo o nullo, il valore assoluto **non cambia nulla** ($|5|=5$). Se $x$ è negativo, il valore assoluto lo rende positivo cambiandogli segno ($|-5| = -(-5) = 5$). In entrambi i casi il risultato è $\ge 0$.

Nota importante: $-x$ nella seconda riga **non significa "un numero negativo"**. Se $x=-5$ (negativo), allora $-x = -(-5) = 5$ (positivo!). Il simbolo $-x$ significa "l'opposto di $x$", che è positivo proprio quando $x$ è negativo.

### Proprietà fondamentali

Dalla definizione derivano proprietà che useremo spessissimo nei moduli successivi:

- $|x| \ge 0$ sempre, e $|x|=0$ solo se $x=0$.
- $|{-x}| = |x|$: numeri opposti hanno lo stesso valore assoluto (simmetria rispetto a 0).
- $|x \cdot y| = |x|\cdot|y|$: il valore assoluto di un prodotto è il prodotto dei valori assoluti.
- **Disuguaglianza triangolare**: $|x+y| \le |x|+|y|$. Il nome viene dal fatto che, pensando a $x$ e $y$ come "spostamenti", la distanza diretta da inizio a fine ($|x+y|$) non può mai superare la somma dei singoli spostamenti ($|x|+|y|$) — un po' come in un triangolo un lato non può superare la somma degli altri due.

### Valore assoluto e distanza tra due punti

Il valore assoluto ci permette anche di misurare la distanza tra due numeri qualsiasi sulla retta, non solo la distanza da zero:

$$\text{distanza tra } a \text{ e } b = |a-b|$$

Per esempio, la distanza tra $3$ e $7$ è $|3-7|=|-4|=4$, la stessa che otterremmo calcolando $|7-3|=4$: l'ordine non conta, perché $|a-b|=|b-a|$ (conseguenza della proprietà $|{-x}|=|x|$, con $x=a-b$).

### Equazioni e disequazioni con il valore assoluto (un primo assaggio)

Se $|x|=3$, stiamo chiedendo: quali numeri distano $3$ da zero? Ce ne sono due: $x=3$ oppure $x=-3$.

Se $|x|<3$, stiamo chiedendo: quali numeri distano meno di $3$ da zero? Tutti quelli tra $-3$ e $3$: $-3<x<3$.

Approfondiremo equazioni e disequazioni con il valore assoluto nel Modulo 1 (notebook 1.14 e 1.15): qui ci basta l'intuizione di base.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# Valore assoluto con SymPy si scrive Abs
print("|5| =", sp.Abs(5))
print("|-5| =", sp.Abs(-5))

# Verifica delle proprietà su alcuni esempi
a, b = -3, 7
print(f"\n|{a}*{b}| = {abs(a*b)}   |{a}|*|{b}| = {abs(a)*abs(b)}   uguali: {abs(a*b) == abs(a)*abs(b)}")

# Disuguaglianza triangolare: verifichiamola su vari esempi (non è una dimostrazione, solo un controllo)
casi = [(-3, 7), (2, 5), (-4, -6), (0, 9)]
for x_val, y_val in casi:
    sinistra = abs(x_val + y_val)
    destra = abs(x_val) + abs(y_val)
    print(f"|{x_val}+{y_val}|={sinistra}  <=  |{x_val}|+|{y_val}|={destra}  ok: {sinistra <= destra}")

# Distanza tra due numeri
distanza = sp.Abs(3 - 7)
print(f"\nDistanza tra 3 e 7: {distanza}")
````

## Esercizio svolto e commentato

**Problema:** risolvi $|x-2|=5$, spiegando il ragionamento.

**Ragionamento:**

L'equazione $|x-2|=5$ dice: "il numero $x-2$ dista $5$ da zero". Sappiamo dalla teoria che questo succede in esattamente due casi: quando $x-2$ vale $5$, oppure quando $x-2$ vale $-5$ (i due numeri che distano 5 da zero).

Trasformiamo quindi l'equazione con valore assoluto in **due equazioni separate, senza valore assoluto**:

$$x - 2 = 5 \qquad \text{oppure} \qquad x-2=-5$$

Risolviamo la prima: $x = 5+2 = 7$.

Risolviamo la seconda: $x = -5+2 = -3$.

Quindi le soluzioni sono $x=7$ e $x=-3$.

**Verifica del risultato (sempre utile con il valore assoluto, perché è facile sbagliare segno):**

- Per $x=7$: $|7-2|=|5|=5$. ✓
- Per $x=-3$: $|-3-2|=|-5|=5$. ✓

**Perché "sdoppiare" l'equazione è il modo corretto di procedere:** il valore assoluto nasconde un'informazione (il segno), quindi per "disfare" $|\cdot|$ dobbiamo esplicitamente considerare entrambe le possibilità che potevano dare origine allo stesso valore assoluto. Ignorare uno dei due casi significherebbe perdere una soluzione valida.

````python
import sympy as sp

x = sp.symbols('x', real=True)
soluzioni = sp.solve(sp.Eq(sp.Abs(x - 2), 5), x)
print("Soluzioni di |x-2|=5:", soluzioni)
````

## Riepilogo finale

- $|x|$ rappresenta la "distanza da zero" di $x$: è sempre $\ge 0$.
- Definizione: $|x|=x$ se $x\ge0$; $|x|=-x$ se $x<0$ (dove $-x$ è l'opposto, non necessariamente un numero negativo).
- $|a-b|$ è la distanza tra due numeri $a$ e $b$ sulla retta numerica.
- Un'equazione $|A|=k$ (con $k>0$) si risolve sempre "sdoppiandola" in due equazioni: $A=k$ oppure $A=-k$.
- Proprietà chiave da ricordare: $|xy|=|x||y|$ e la disuguaglianza triangolare $|x+y|\le|x|+|y|$.
