## Perché ci interessa

Ogni volta che diciamo "se piove, allora prendo l'ombrello" o "il numero è pari e maggiore di 10", stiamo combinando delle affermazioni con delle regole precise. In matematica queste regole non possono essere lasciate all'intuizione: servono per costruire dimostrazioni corrette, capire teoremi, e non farsi ingannare da ragionamenti che *sembrano* giusti ma non lo sono.

La **logica proposizionale** è lo studio di come combinare affermazioni vere o false. È lo scheletro su cui si costruisce ogni dimostrazione matematica rigorosa, che vedremo sempre di più da qui fino al Modulo 10.

## Teoria, passo dopo passo

### Cos'è una proposizione

Una **proposizione** è un'affermazione a cui si può assegnare senza ambiguità un valore di verità: **vero** (V) oppure **falso** (F), mai entrambi, mai nessuno dei due.

- "$7$ è un numero primo" → proposizione, vera.
- "$4$ è maggiore di $10$" → proposizione, falsa.
- "Che ore sono?" → **non** è una proposizione (è una domanda, non ha valore di verità).
- "$x+1=5$" → **non** è una proposizione finché $x$ non è specificato (dipende dal valore di $x$): diventerà una proposizione con l'aiuto dei quantificatori, che vedremo nel notebook 0.17.

Indichiamo le proposizioni con lettere minuscole: $p$, $q$, $r$...

### I connettivi logici principali

Possiamo combinare proposizioni con dei **connettivi**, ognuno con un significato preciso, riassunto in una **tavola di verità** (che elenca tutti i casi possibili).

**Negazione** ($\neg p$, si legge "non $p$"): vera quando $p$ è falsa, e viceversa.

| $p$ | $\neg p$ |
|---|---|
| V | F |
| F | V |

**Congiunzione** ($p \wedge q$, "$p$ e $q$"): vera solo quando **entrambe** sono vere.

| $p$ | $q$ | $p \wedge q$ |
|---|---|---|
| V | V | V |
| V | F | F |
| F | V | F |
| F | F | F |

**Disgiunzione** ($p \vee q$, "$p$ oppure $q$"): vera quando **almeno una** delle due è vera (attenzione: è un "oppure inclusivo", vera anche se sono vere entrambe — diverso dall'"aut...aut" del linguaggio comune).

| $p$ | $q$ | $p \vee q$ |
|---|---|---|
| V | V | V |
| V | F | V |
| F | V | V |
| F | F | F |

### Il collegamento con gli insiemi (notebook 0.13)

Non è un caso che queste tavole di verità assomiglino alle operazioni tra insiemi già viste: "$e$" ($\wedge$) si comporta come l'intersezione ($\cap$), "$oppure$" ($\vee$) si comporta come l'unione ($\cup$), la negazione si comporta come il complementare. Non è una coincidenza: entrambe le teorie condividono la stessa struttura logica di fondo. Ritroveremo infatti le leggi di De Morgan, viste per gli insiemi, identiche anche in logica:

$$\neg(p \wedge q) \equiv (\neg p) \vee (\neg q) \qquad \neg(p \vee q) \equiv (\neg p) \wedge (\neg q)$$

(il simbolo $\equiv$ significa "logicamente equivalente": le due proposizioni hanno sempre lo stesso valore di verità, qualunque siano $p$ e $q$).

````python
import itertools

def negazione(p): return not p
def congiunzione(p, q): return p and q
def disgiunzione(p, q): return p or q

# Ricostruiamo le tavole di verità con Python
print("Tavola di verità di p AND q, p OR q:")
print(f"{'p':^5}{'q':^5}{'p AND q':^10}{'p OR q':^10}")
for p, q in itertools.product([True, False], repeat=2):
    print(f"{str(p):^5}{str(q):^5}{str(congiunzione(p,q)):^10}{str(disgiunzione(p,q)):^10}")

# Verifica delle leggi di De Morgan su tutti i casi possibili
print("\nVerifica leggi di De Morgan (su tutte le combinazioni V/F):")
tutte_valide = True
for p, q in itertools.product([True, False], repeat=2):
    legge1 = (negazione(congiunzione(p, q)) == disgiunzione(negazione(p), negazione(q)))
    legge2 = (negazione(disgiunzione(p, q)) == congiunzione(negazione(p), negazione(q)))
    tutte_valide = tutte_valide and legge1 and legge2
print("Le leggi di De Morgan valgono in tutti i casi?", tutte_valide)
````

## Esercizio svolto e commentato

**Problema:** costruisci la tavola di verità di $(p \wedge q) \vee (\neg p)$, e osserva cosa succede.

**Ragionamento:**

Costruiamo la tavola passo per passo, calcolando prima le sotto-espressioni più semplici e poi combinandole, esattamente come faremmo con l'ordine delle operazioni aritmetiche (notebook 0.7): prima $\neg p$, poi $p\wedge q$, infine la disgiunzione tra i due risultati.

| $p$ | $q$ | $\neg p$ | $p\wedge q$ | $(p\wedge q)\vee(\neg p)$ |
|---|---|---|---|---|
| V | V | F | V | V |
| V | F | F | F | F |
| F | V | V | F | V |
| F | F | V | F | V |

Guardiamo il risultato finale: è vero in tutti i casi **tranne** quando $p$ è vera e $q$ è falsa. Osserviamo che questo è esattamente lo stesso schema di verità dell'implicazione $p \Rightarrow q$ (che studieremo nel dettaglio nel prossimo notebook, 0.16): l'unico caso falso è "$p$ vera, $q$ falsa".

**Perché costruire la tavola riga per riga, colonna per colonna, è un metodo affidabile:** con proposizioni complesse è facile "perdersi" cercando di valutare tutto a mente. Scomponendo l'espressione nei suoi pezzi più semplici, e costruendo una colonna alla volta, si riduce il rischio di errore — lo stesso principio dell'ordine delle operazioni: risolvere un pezzo alla volta, senza saltare passaggi.

````python
import itertools

print(f"{'p':^5}{'q':^5}{'not p':^7}{'p and q':^9}{'(p and q) or (not p)':^22}")
for p, q in itertools.product([True, False], repeat=2):
    not_p = not p
    p_and_q = p and q
    risultato = p_and_q or not_p
    print(f"{str(p):^5}{str(q):^5}{str(not_p):^7}{str(p_and_q):^9}{str(risultato):^22}")
````

## Riepilogo finale

- Una proposizione è un'affermazione con valore di verità univoco (vera o falsa, mai entrambe).
- La negazione ($\neg$) inverte il valore di verità; la congiunzione ($\wedge$, "e") è vera solo se entrambe lo sono; la disgiunzione ($\vee$, "oppure") è vera se almeno una lo è.
- Le tavole di verità elencano tutti i casi possibili e permettono di valutare qualsiasi espressione logica in modo sistematico, un pezzo alla volta.
- Le leggi di De Morgan collegano negazione, congiunzione e disgiunzione: hanno esattamente la stessa struttura viste per gli insiemi nel notebook 0.13 (non è una coincidenza).
- Due proposizioni composte sono logicamente equivalenti se hanno la stessa tavola di verità, indipendentemente da come sono scritte.
