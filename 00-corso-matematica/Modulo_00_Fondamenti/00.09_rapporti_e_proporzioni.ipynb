## Perché ci interessa

Spesso non ci interessa la quantità assoluta di qualcosa, ma il **confronto** tra due quantità: "quante volte" una grandezza è più grande di un'altra. Se una ricetta usa 2 uova ogni 3 tazze di farina, e vuoi farne una versione più grande mantenendo lo stesso sapore, devi mantenere quello stesso confronto, non le stesse quantità assolute.

Questa idea di confronto si formalizza con i **rapporti**, e l'uguaglianza tra due rapporti si chiama **proporzione**. Sono strumenti che ritroverai ovunque: dalle scale delle cartine geografiche, alla chimica (concentrazioni), alla geometria (triangoli simili, che vedremo nel Modulo 2).

## Teoria, passo dopo passo

### Cos'è un rapporto

Il **rapporto** tra due grandezze $a$ e $b$ (con $b \ne 0$) è semplicemente la loro divisione, scritta come frazione:

$$a : b = \frac{a}{b}$$

Per esempio, se in una classe ci sono 12 femmine e 8 maschi, il rapporto femmine:maschi è $12:8 = \frac{12}{8} = \frac{3}{2}$: ogni 3 femmine ci sono 2 maschi. Nota che, esattamente come per le frazioni (notebook 0.3), un rapporto si può sempre semplificare ai minimi termini senza cambiare il suo significato.

### Cos'è una proporzione

Una **proporzione** è l'uguaglianza tra due rapporti:

$$a : b = c : d \qquad \text{cioè} \qquad \frac{a}{b} = \frac{c}{d}$$

Si legge "$a$ sta a $b$ come $c$ sta a $d$". I numeri $a$ e $d$ si chiamano **estremi**, i numeri $b$ e $c$ si chiamano **medi**.

### La proprietà fondamentale delle proporzioni

Da $\frac{a}{b}=\frac{c}{d}$, moltiplicando entrambi i lati per $bd$ (lecito perché $b,d\ne0$), otteniamo:

$$a \cdot d = b \cdot c$$

**Il prodotto degli estremi è uguale al prodotto dei medi.** Questa non è una regola aggiuntiva da ricordare a memoria: è la stessa condizione di equivalenza tra frazioni che abbiamo già visto nel notebook 0.3! Una proporzione, in fondo, è solo un'uguaglianza tra due frazioni scritta con un'altra notazione.

Questa proprietà è utilissima per trovare un termine sconosciuto in una proporzione. Se conosciamo $a, b, c$ e vogliamo trovare $d$:

$$a \cdot d = b \cdot c \quad \Rightarrow \quad d = \frac{b \cdot c}{a}$$

Questo procedimento si chiama, in alcuni libri, "regola del tre" (o "terzo proporzionale" quando si cerca $d$ note le altre tre grandezze).

### Grandezze direttamente e inversamente proporzionali

Due grandezze $x$ e $y$ sono **direttamente proporzionali** se il loro rapporto è costante:

$$\frac{y}{x} = k \quad \text{(costante)} \qquad \Longleftrightarrow \qquad y = kx$$

Esempio: il costo totale di alcuni oggetti identici è direttamente proporzionale al numero di oggetti comprati (se ne compri il doppio, spendi il doppio).

Due grandezze $x$ e $y$ sono **inversamente proporzionali** se il loro **prodotto** è costante:

$$x \cdot y = k \quad \text{(costante)} \qquad \Longleftrightarrow \qquad y = \frac{k}{x}$$

Esempio: se devi percorrere una distanza fissa, il tempo impiegato e la velocità sono inversamente proporzionali (raddoppiando la velocità, il tempo si dimezza, ma il loro prodotto — la distanza — resta lo stesso).

````python
import sympy as sp

# Verifichiamo la proprietà fondamentale delle proporzioni
a, b, c, d = 3, 4, 6, 8  # 3:4 = 6:8 ?
proporzione_valida = (a * d == b * c)
print(f"{a}:{b} = {c}:{d}  è una proporzione valida? {proporzione_valida}")  # 3*8=24, 4*6=24 -> True

# Troviamo un termine sconosciuto: 5 : 7 = x : 21
a, b, c = 5, 7, 21
d = sp.Rational(b * c, a)
print(f"\n5 : 7 = x : 21  ->  x = {d}")

# Grandezze direttamente proporzionali: costo totale vs numero di oggetti
prezzo_unitario = 3  # euro per oggetto
numeri_oggetti = [1, 2, 3, 4, 5]
costi = [prezzo_unitario * n for n in numeri_oggetti]
print("\nOggetti:", numeri_oggetti)
print("Costi:  ", costi)
print("Rapporto costo/numero (deve essere sempre uguale):", [sp.Rational(c, n) for c, n in zip(costi, numeri_oggetti)])

# Grandezze inversamente proporzionali: velocità vs tempo, distanza fissa = 120 km
distanza = 120
velocità = [20, 30, 40, 60]
tempi = [sp.Rational(distanza, v) for v in velocità]
print("\nVelocità (km/h):", velocità)
print("Tempi (ore):     ", tempi)
print("Prodotto velocità*tempo (deve essere sempre 120):", [v * t for v, t in zip(velocità, tempi)])
````

## Esercizio svolto e commentato

**Problema:** una ricetta per 4 persone richiede 250 g di farina. Quanta farina serve per 6 persone?

**Ragionamento:**

Il rapporto tra farina e numero di persone deve restare costante (è una situazione di proporzionalità diretta: raddoppiando le persone, serve il doppio di farina). Impostiamo la proporzione:

$$\text{farina per 4 persone} : \text{4 persone} = \text{farina per 6 persone} : \text{6 persone}$$

$$250 : 4 = x : 6$$

Usiamo la proprietà fondamentale (prodotto degli estremi = prodotto dei medi):

$$250 \times 6 = 4 \times x$$

$$1500 = 4x$$

$$x = \frac{1500}{4} = 375$$

Servono **375 grammi** di farina.

**Perché impostare la proporzione in questo modo (e non, per esempio, persone:farina = persone:farina) funziona comunque, purché si sia coerenti:** l'importante è mettere sempre la stessa grandezza nella stessa posizione (qui: farina sempre a sinistra dei due punti, persone sempre a destra). Il risultato non cambia se si scambiano coerentemente entrambe le posizioni, ma cambierebbe (e sarebbe sbagliato) se si mischiassero le posizioni a metà.

````python
import sympy as sp

x = sp.symbols('x', positive=True)
soluzione = sp.solve(sp.Eq(sp.Rational(250, 4), x / 6), x)
print("Farina necessaria per 6 persone:", soluzione, "grammi")
````

**1.** $8 \times 15 = 5 \times x \Rightarrow 120 = 5x \Rightarrow x=24$.

**2.** $1 \text{ cm} : 50000 \text{ cm} = 8 \text{ cm} : x$. Quindi $x = 8 \times 50000 = 400000$ cm $= 4000$ m $= 4$ km.

Sviluppiamo il lato destro: $b(c+d)=bc+bd$. Poiché sappiamo che $ad=bc$, i due lati diventano $ad+bd$ e $bc+bd=ad+bd$: sono uguali! Quindi la proporzione $(a+b):b=(c+d):d$ è verificata.

Verifica computazionale:
```python
import sympy as sp

a, b, c, d = 3, 4, 6, 8  # 3:4 = 6:8
sinistra = sp.Rational(a + b, b)
destra = sp.Rational(c + d, d)
print(f"(a+b):b = {sinistra}   (c+d):d = {destra}   uguali: {sinistra == destra}")
```

## Riepilogo finale

- Un rapporto $a:b$ è semplicemente la frazione $\frac{a}{b}$; una proporzione è l'uguaglianza tra due rapporti.
- Proprietà fondamentale: in $a:b=c:d$, il prodotto degli estremi ($a\cdot d$) è uguale al prodotto dei medi ($b\cdot c$) — è la stessa condizione di equivalenza tra frazioni vista nel notebook 0.3.
- Due grandezze sono **direttamente proporzionali** se il loro rapporto è costante ($y=kx$); sono **inversamente proporzionali** se il loro prodotto è costante ($y=k/x$).
- Riconoscere se una situazione è di proporzionalità diretta o inversa è il passo più importante (e più delicato) prima di impostare i calcoli.
