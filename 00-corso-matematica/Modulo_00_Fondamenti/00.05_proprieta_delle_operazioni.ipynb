## Teoria, passo dopo passo

### Proprietà commutativa

Per addizione e moltiplicazione, l'ordine degli operandi non conta:

$$a+b=b+a \qquad a\times b = b\times a$$

Esempio: $3+5=5+3=8$. **Attenzione**: sottrazione e divisione **non** sono commutative: $5-3=2$ ma $3-5=-2$ (diverso); $10\div2=5$ ma $2\div10=0.2$ (diverso).

### Proprietà associativa

Per addizione e moltiplicazione, quando ci sono tre o più termini, non importa come li raggruppiamo con le parentesi:

$$(a+b)+c = a+(b+c) \qquad (a\times b)\times c = a\times(b\times c)$$

Questa è la proprietà che abbiamo usato "senza dirlo" nel notebook 0.2, per riorganizzare $(a+b)+((-a)+(-b))$ nell'esercizio svolto. Grazie all'associativa possiamo scrivere semplicemente $a+b+c$ senza parentesi: il risultato non cambia comunque le raggruppiamo. Anche qui, sottrazione e divisione **non** sono associative: $(10-3)-2=5$ ma $10-(3-2)=9$ (diverso).

### Elemento neutro ed elemento opposto/reciproco

Lo **0** è l'elemento neutro dell'addizione: $a+0=a$ per ogni $a$. Il numero **1** è l'elemento neutro della moltiplicazione: $a\times1=a$ per ogni $a$.

Ogni numero $a$ ha un **opposto** $-a$, tale che $a+(-a)=0$ (lo abbiamo definito nel notebook 0.2). Ogni numero $a\ne0$ ha un **reciproco** $\frac1a$, tale che $a\times\frac1a=1$ (lo abbiamo usato nel notebook 0.3 per definire la divisione tra frazioni come moltiplicazione per il reciproco).

### Proprietà distributiva

La distributiva collega addizione e moltiplicazione:

$$a\times(b+c) = a\times b + a\times c$$

Perché è vera? Pensiamo alla moltiplicazione come somma ripetuta (notebook 0.6, che vedremo tra poco, motiverà questo collegamento in modo più formale): "$a$ ripetuto $(b+c)$ volte" è lo stesso di "$a$ ripetuto $b$ volte, più $a$ ripetuto $c$ volte". Usiamo la distributiva in due direzioni:

- **Da sinistra a destra** ("espandere"): $3\times(x+2) = 3x+6$.
- **Da destra a sinistra** ("raccogliere a fattor comune"): $5x+5y = 5(x+y)$. È esattamente quello che abbiamo fatto nel notebook 0.1, scrivendo $2k_1+2k_2=2(k_1+k_2)$ per dimostrare che la somma di due pari è pari.

### Proprietà invariantiva: il fondamento per risolvere equazioni

Questa proprietà riguarda le **uguaglianze**, non le singole espressioni, e dice che possiamo modificare entrambi i membri di un'uguaglianza in certi modi senza romperla:

**Invariantiva dell'addizione**: se $a=b$, allora $a+c=b+c$ per qualunque $c$ (possiamo sommare o sottrarre lo stesso numero a entrambi i membri).

**Invariantiva della moltiplicazione**: se $a=b$, allora $a\times c=b\times c$ per qualunque $c\ne0$ (possiamo moltiplicare o dividere entrambi i membri per lo stesso numero non nullo — il motivo per cui $c\ne0$ lo capiremo studiando le equazioni fratte nel Modulo 1).

**Perché funziona:** se $a$ e $b$ sono davvero lo stesso numero, aggiungere (o moltiplicare per) $c$ a entrambi produce ancora lo stesso numero da entrambe le parti — è quasi un'osservazione ovvia, ma è precisamente ciò che rende possibile "spostare" termini da un lato all'altro di un'equazione: quando scrivi $x+5=12$ e concludi $x=12-5$, stai applicando l'invariantiva dell'addizione con $c=-5$ a entrambi i membri. Non è un "trucco": è una conseguenza diretta di questa proprietà.

````python
import sympy as sp

a, b, c, x = sp.symbols('a b c x')

# Verifica simbolica delle proprietà (SymPy conferma che le due forme sono sempre uguali)
print("Commutativa somma:", sp.simplify((a + b) - (b + a)) == 0)
print("Associativa prodotto:", sp.simplify(((a * b) * c) - (a * (b * c))) == 0)
print("Distributiva:", sp.expand(a * (b + c)) == sp.expand(a * b + a * c))

# Raccogliere a fattor comune: SymPy lo fa con factor()
espressione = 5 * x + 5 * 3
print("\n5x + 15 raccolto:", sp.factor(espressione))  # atteso 5*(x+3)

# Proprietà invariantiva applicata per isolare x in x + 5 = 12
equazione = sp.Eq(x + 5, 12)
# sommiamo -5 a entrambi i membri (invariantiva dell'addizione)
nuova_equazione = sp.Eq(equazione.lhs - 5, equazione.rhs - 5)
print("\nEquazione originale:", equazione)
print("Dopo invariantiva (-5 a entrambi i membri):", sp.simplify(nuova_equazione))
````

## Esercizio svolto e commentato

**Problema:** risolvi $2(x+3)=14$, usando esplicitamente la distributiva e la proprietà invariantiva, spiegando ogni passaggio.

**Ragionamento:**

**Passo 1 — applichiamo la distributiva per eliminare la parentesi:**

$$2(x+3) = 2x+6$$

L'equazione diventa $2x+6=14$.

**Passo 2 — applichiamo l'invariantiva dell'addizione, sottraendo $6$ da entrambi i membri:**

$$2x+6-6 = 14-6 \quad \Rightarrow \quad 2x = 8$$

Sottraendo lo stesso numero da entrambe le parti, l'uguaglianza resta valida (per definizione della proprietà invariantiva), e a sinistra $6-6=0$ fa "sparire" il termine costante, isolando il termine con $x$.

**Passo 3 — applichiamo l'invariantiva della moltiplicazione, dividendo entrambi i membri per $2$ (lecito perché $2\ne0$):**

$$\frac{2x}{2} = \frac{8}{2} \quad \Rightarrow \quad x=4$$

**Verifica** (sostituendo $x=4$ nell'equazione originale): $2(4+3)=2\times7=14$. ✓

**Perché ogni passaggio è giustificato e non è "spostare i numeri dall'altra parte cambiando segno" come si dice spesso a memoria:** quello che sembra un trucco meccanico ("porto il 6 dall'altra parte e cambio segno") è in realtà l'applicazione dell'invariantiva: stiamo sottraendo la stessa quantità da entrambi i membri, e il risultato *sembra* uno spostamento con cambio di segno solo perché $+6-6$ si annulla a sinistra e $-6$ resta a destra. Capire questo meccanismo, invece di applicarlo a memoria, sarà indispensabile per non confondersi con equazioni più complesse nel Modulo 1.

````python
import sympy as sp

x = sp.symbols('x')
soluzione = sp.solve(sp.Eq(2 * (x + 3), 14), x)
print("Soluzione:", soluzione)

# Verifica sostituendo il valore trovato nell'equazione originale
verifica = 2 * (soluzione[0] + 3)
print("Verifica: 2*(x+3) con x=", soluzione[0], "dà", verifica, "(atteso 14)")
````

Questo succede perché la distributiva collega moltiplicazione e addizione ($a\times(b+c)=ab+ac$), ma **non** esiste una proprietà analoga per la divisione quando la somma è al denominatore: dividere per $(a+b)$ non è la stessa cosa che dividere separatamente per $a$ e per $b$. La distributiva vale invece se la somma è al *numeratore*: $\frac{a+b}{c}=\frac{a}{c}+\frac{b}{c}$ è corretta, perché equivale a moltiplicare $(a+b)$ per il reciproco $\frac{1}{c}$, e lì la distributiva si applica regolarmente.

destra = sp.Rational(c, a) + sp.Rational(c, b)

print(f"c/a + c/b = {destra}")

## Riepilogo finale

- **Commutativa** (somma/prodotto): l'ordine degli operandi non conta. **Associativa** (somma/prodotto): il raggruppamento con le parentesi non conta. Nessuna delle due vale per sottrazione o divisione.
- **0** è l'elemento neutro dell'addizione, **1** quello della moltiplicazione; ogni numero ha un opposto (somma 0) e, se diverso da 0, un reciproco (prodotto 1).
- **Distributiva**: $a(b+c)=ab+ac$, usabile sia per espandere sia per raccogliere a fattor comune.
- **Invariantiva**: sommare/sottrarre uno stesso numero, o moltiplicare/dividere per uno stesso numero non nullo, a entrambi i membri di un'uguaglianza la lascia valida — è il meccanismo dietro ogni tecnica di risoluzione di equazioni.
- Queste proprietà non sono regole a memoria: derivano da come sono costruiti i numeri, e le useremo esplicitamente da qui in avanti, a partire dal prossimo notebook sulle potenze.
