## Perché ci interessa

Molto spesso in matematica non ci interessa un singolo numero, ma **tutti i numeri reali compresi tra due valori**: per esempio, "tutte le temperature tra 0 e 100 gradi" oppure, più avanti nel corso, "tutti gli $x$ per cui una funzione è definita".

Gli **intervalli** sono il modo standard per descrivere questi sottoinsiemi "continui" di $\mathbb{R}$. Sono un caso specifico degli insiemi visti nei notebook precedenti, ma abbastanza importanti e ricorrenti da meritare una notazione dedicata.

## Teoria, passo dopo passo

### Intervalli limitati

Dati due numeri reali $a < b$, possiamo definire quattro tipi di intervallo, a seconda che gli estremi $a$ e $b$ siano inclusi o no:

- **Intervallo chiuso** $[a,b] = \{x \in \mathbb{R} : a \le x \le b\}$ — entrambi gli estremi inclusi.
- **Intervallo aperto** $(a,b) = \{x \in \mathbb{R} : a < x < b\}$ — entrambi gli estremi esclusi.
- **Intervallo semiaperto a destra** $[a,b) = \{x \in \mathbb{R} : a \le x < b\}$ — $a$ incluso, $b$ escluso.
- **Intervallo semiaperto a sinistra** $(a,b] = \{x \in \mathbb{R} : a < x \le b\}$ — $a$ escluso, $b$ incluso.

La parentesi **quadra** significa "estremo incluso", la parentesi **tonda** significa "estremo escluso". Alcuni libri usano invece $]a,b[$ al posto di $(a,b)$: è solo una notazione alternativa, il significato è identico.

**Perché la distinzione tra incluso/escluso è importante:** $[0,1]$ contiene $1$, mentre $[0,1)$ no. Sembra un dettaglio minore, ma diventerà fondamentale quando parleremo di massimi e minimi (Modulo 5): una funzione può avere massimo su $[0,1]$ (raggiunto proprio in $x=1$) ma non su $[0,1)$ (il valore in $1$ non fa parte del dominio, anche se ci si avvicina sempre di più).

### Intervalli illimitati

Possiamo anche descrivere insiemi che si estendono all'infinito in una direzione, usando i simboli $+\infty$ e $-\infty$ (che **non sono numeri reali**, sono solo simboli per indicare "senza limite"):

- $[a, +\infty) = \{x\in\mathbb{R} : x \ge a\}$
- $(a, +\infty) = \{x\in\mathbb{R} : x > a\}$
- $(-\infty, b] = \{x\in\mathbb{R} : x \le b\}$
- $(-\infty, b) = \{x\in\mathbb{R} : x < b\}$
- $(-\infty, +\infty) = \mathbb{R}$ (tutta la retta reale)

Nota che accanto a $+\infty$ e $-\infty$ si usa **sempre** la parentesi tonda (esclusa): non ha senso dire che l'infinito è "incluso", perché non è un numero raggiungibile.

### Operazioni tra intervalli

Gli intervalli sono insiemi, quindi possiamo applicare le operazioni viste nel notebook 0.13. Un caso molto comune è l'**intersezione**, che serve per trovare i valori che soddisfano più condizioni contemporaneamente. Esempio:

$$[0, 5] \cap [3, 8] = [3, 5]$$

perché solo i numeri tra 3 e 5 stanno in entrambi gli intervalli. Anche l'**unione** è comune: se gli intervalli non si sovrappongono, l'unione non è più un unico intervallo, ma una **collezione di intervalli disgiunti**:

$$[0,2] \cup [5,7]$$

resta scritto così, non si può semplificare in un unico intervallo, perché c'è un "vuoto" tra 2 e 5.

````python
import sympy as sp

# SymPy rappresenta gli intervalli con sp.Interval
I1 = sp.Interval(0, 5)          # [0,5] chiuso
I2 = sp.Interval(3, 8)          # [3,8] chiuso
I3 = sp.Interval.open(0, 1)     # (0,1) aperto
I4 = sp.Interval(0, 1, False, True)  # [0,1) semiaperto a destra (left_open=False, right_open=True)

print("I1 =", I1)
print("I3 (aperto) =", I3)
print("I4 (semiaperto a destra) =", I4)

# Intersezione: valori che soddisfano entrambe le condizioni
intersezione = I1.intersect(I2)
print("\n[0,5] ∩ [3,8] =", intersezione)

# Unione di intervalli disgiunti: resta una collezione, non un unico intervallo
unione_disgiunta = sp.Interval(0, 2).union(sp.Interval(5, 7))
print("[0,2] ∪ [5,7] =", unione_disgiunta)

# Intervalli illimitati
semiretta = sp.Interval(2, sp.oo, False, True)  # [2, +infinito)
print("\n[2, +infinito) =", semiretta)
print("5 appartiene?", semiretta.contains(5))
print("-1 appartiene?", semiretta.contains(-1))
````

## Esercizio svolto e commentato

**Problema:** trova, come intervallo, l'insieme dei numeri reali che soddisfano **entrambe** le condizioni $x > -2$ e $x \le 4$.

**Ragionamento:**

Le due condizioni definiscono, separatamente, due sottoinsiemi di $\mathbb{R}$:

- $x > -2$ corrisponde all'intervallo $(-2, +\infty)$.
- $x \le 4$ corrisponde all'intervallo $(-\infty, 4]$.

Volere **entrambe** le condizioni contemporaneamente significa cercare l'**intersezione** dei due insiemi (esattamente come nel notebook 0.13, dove "e" corrispondeva a $\cap$):

$$(-2, +\infty) \cap (-\infty, 4]$$

Per trovare l'intersezione, confrontiamo gli estremi: il "sinistro" più restrittivo è $-2$ (escluso, perché viene dal primo intervallo, che lo esclude), il "destro" più restrittivo è $4$ (incluso, perché viene dal secondo intervallo, che lo include). Non c'è conflitto tra le due condizioni (l'intervallo risultante non è vuoto, perché $-2 < 4$), quindi:

$$(-2, +\infty) \cap (-\infty, 4] = (-2, 4]$$

**Perché guardiamo separatamente ogni estremo (incluso/escluso) invece di "unire" genericamente le due notazioni:** ogni condizione porta con sé la propria informazione su inclusione/esclusione. Dobbiamo sempre riportare fedelmente questa informazione nel risultato finale: sarebbe un errore scrivere, per esempio, $[-2,4]$ (che includerebbe erroneamente $-2$, mentre la condizione originale $x>-2$ lo esclude).

````python
import sympy as sp

I1 = sp.Interval(-2, sp.oo, True, True)   # (-2, +infinito), aperto a sinistra
I2 = sp.Interval(-sp.oo, 4, True, False)  # (-infinito, 4], chiuso a destra

risultato = I1.intersect(I2)
print("Risultato:", risultato)  # atteso (-2, 4]
````

## Riepilogo finale

- Un intervallo descrive tutti i numeri reali compresi tra due estremi; la parentesi quadra $[\ ]$ include l'estremo, la tonda $(\ )$ lo esclude.
- $+\infty$ e $-\infty$ non sono numeri: indicano solo "senza limite", e vanno sempre scritti con parentesi tonda.
- L'intersezione di intervalli corrisponde a richiedere **entrambe** le condizioni; l'unione corrisponde a richiederne **almeno una**.
- L'unione di due intervalli che non si sovrappongono (e non sono "adiacenti senza buchi") non è più un unico intervallo: resta una collezione di intervalli disgiunti.
- Nel confrontare/combinare intervalli, bisogna sempre tracciare con attenzione se ogni estremo è incluso o escluso.
