## Perché ci interessa

Il notebook 18.5 ha mostrato che i numeri primi sono gli "atomi" moltiplicativi di $\mathbb Z$ (teorema fondamentale dell'aritmetica), e il notebook 18.6 ha mostrato che sono esattamente i moduli $n$ per cui $\mathbb Z/n\mathbb Z$ è un campo. Due ruoli fondamentali per una domanda che a questo punto è inevitabile: **quanti sono** i numeri primi, e **come sono distribuiti** tra gli interi?

La prima domanda ha una risposta sorprendentemente semplice e dimostrata già più di duemila anni fa (Euclide). La seconda è uno dei problemi più profondi e ancora attivi della matematica: la accenneremo senza poterla approfondire, ma vale la pena sapere cosa si nasconde dietro la parola "primo" oltre alla definizione elementare.


## Teoria, passo dopo passo

### Definizione

Un intero $p>1$ è **primo** se i suoi unici divisori positivi sono $1$ e $p$ stesso. Un intero $>1$ che non è primo si dice **composto** (per il teorema fondamentale dell'aritmetica, notebook 18.5, ogni composto si fattorizza in primi più piccoli). Per convenzione, $1$ non è né primo né composto (se lo fosse, l'unicità della fattorizzazione del notebook 18.5 sarebbe falsa: $6=2\cdot3=1\cdot2\cdot3=1\cdot1\cdot2\cdot3=\ldots$, infinite fattorizzazioni).

### Il teorema di Euclide: infiniti numeri primi

**Teorema.** Esistono infiniti numeri primi.

**Dimostrazione (per assurdo, notebook 10.6).** Supponiamo che ne esistano solo un numero finito, $p_1,p_2,\ldots,p_k$ (l'intera lista). Consideriamo
$$N=p_1p_2\cdots p_k+1.$$
$N>1$, quindi per il teorema fondamentale dell'aritmetica (notebook 18.5) $N$ ha almeno un fattore primo $q$. Per ipotesi, $q$ dev'essere uno dei $p_i$ della lista (che, per assunzione, li contiene tutti). Ma allora $q=p_i$ divide sia $N$ sia $p_1p_2\cdots p_k$ (essendo uno dei fattori del prodotto), quindi divide anche la loro differenza:
$$q\mid\big(N-p_1p_2\cdots p_k\big)=1.$$
Ma nessun primo può dividere $1$ (essendo $q\ge2$): contraddizione. L'ipotesi che i primi siano in numero finito è falsa: sono infiniti. $\blacksquare$

**Osservazione.** La dimostrazione non dice che $N$ sia primo (anzi, spesso non lo è: es. $2\cdot3\cdot5\cdot7\cdot11\cdot13+1=30031=59\cdot509$ non è primo); dice solo che $N$ ha *un* fattore primo che non può essere nella lista originale — bastante per la contraddizione, senza bisogno di identificarlo.

### Il crivello di Eratostene

Per **trovare** tutti i primi fino a una soglia $N$, il metodo più antico ed efficiente per calcolo a mano è il **crivello di Eratostene**: si scrivono tutti gli interi da $2$ a $N$; si parte da $p=2$, si eliminano tutti i suoi multipli propri ($4,6,8,\ldots$); si passa al più piccolo numero non ancora eliminato ($3$), si eliminano i suoi multipli propri; si ripete, prendendo ogni volta il più piccolo numero non eliminato, finché $p^2>N$ (a quel punto tutti i numeri composti $\le N$ sono già stati eliminati, perché ogni composto $\le N$ ha un fattore primo $\le\sqrt N$ — se tutti i suoi fattori fossero $>\sqrt N$, il loro prodotto sarebbe $>N$). I numeri rimasti non eliminati sono esattamente i primi $\le N$.

### La distribuzione dei numeri primi

Sia $\pi(x)$ il numero di primi $\le x$ (es. $\pi(10)=4$: $2,3,5,7$). Come cresce $\pi(x)$ al crescere di $x$? Il teorema di Euclide dice solo $\pi(x)\to\infty$, ma non dice **quanto velocemente**.

**Teorema dei numeri primi** (enunciato senza dimostrazione — la dimostrazione, di Hadamard e de la Vallée Poussin nel 1896, usa analisi complessa avanzata, ben oltre lo scopo di questo corso, sebbene costruisca proprio sugli strumenti del Modulo 9):
$$\pi(x)\sim\frac x{\ln x}\qquad\text{per }x\to\infty$$
(nel senso che $\pi(x)/(x/\ln x)\to1$). Intuitivamente: "intorno a un intero grande $x$, la probabilità che un numero scelto a caso vicino a $x$ sia primo è circa $1/\ln x$" — i primi diventano via via più radi, ma mai troppo radi (restano infiniti, come garantito da Euclide).

**Problemi aperti (cenno, senza dimostrazione né approfondimento).** Nonostante duemila anni di studio, alcune domande elementari sui primi restano senza risposta: la **congettura di Goldbach** (ogni intero pari $>2$ è somma di due primi, verificata computazionalmente per numeri enormi ma mai dimostrata in generale) e la **congettura dei primi gemelli** (esistono infinite coppie di primi $p,p+2$, come $(11,13)$ o $(17,19)$). Li citiamo solo per mostrare che, dietro una definizione elementare come "numero primo", si nascondono alcuni dei problemi più difficili di tutta la matematica.


```python
import numpy as np
import matplotlib.pyplot as plt

def crivello_di_eratostene(N):
    # Restituisce l'array dei numeri primi <= N
    e_primo = np.ones(N + 1, dtype=bool)
    e_primo[0:2] = False  # 0 e 1 non sono primi
    for p in range(2, int(N**0.5) + 1):
        if e_primo[p]:
            e_primo[p*p::p] = False  # elimina i multipli propri di p, a partire da p^2
    return np.nonzero(e_primo)[0]

N = 100_000
primi = crivello_di_eratostene(N)
print(f"Numero di primi fino a {N}: {len(primi)}")
print(f"Approssimazione x/ln(x): {N / np.log(N):.0f}")

primi_100 = [p for p in primi if p <= 100]
print(f"\nPrimi fino a 100: {list(primi_100)}")
set_primi_100 = set(primi_100)
gemelli = [(p, p + 2) for p in primi_100 if (p + 2) in set_primi_100]
print(f"Coppie di primi gemelli sotto 100: {gemelli}")

xs = np.array([10**k for k in range(2, 6)])
pi_xs = [np.searchsorted(primi, x, side='right') for x in xs]
approx = xs / np.log(xs)
plt.figure(figsize=(7, 5))
plt.plot(xs, pi_xs, 'o-', label=r'$\pi(x)$ esatto')
plt.plot(xs, approx, 's--', label=r'$x/\ln x$')
plt.xscale('log')
plt.xlabel('x')
plt.legend()
plt.title('Distribuzione dei numeri primi: verifica del teorema dei numeri primi')
plt.show()

for x, pix, ap in zip(xs, pi_xs, approx):
    print(f"x={x:>7}: pi(x)={pix:>6}, x/ln(x)={ap:>8.1f}, rapporto={pix/ap:.3f}")
```

## Esercizio svolto e commentato

**Problema:** verifica numericamente la dimostrazione di Euclide per $k=4$: calcola $N=p_1p_2p_3p_4+1=2\cdot3\cdot5\cdot7+1$, e trova il suo più piccolo fattore primo, verificando che non è tra $2,3,5,7$.

**Ragionamento:**

**Passo 1 — calcoliamo $N$.** $N=2\cdot3\cdot5\cdot7+1=210+1=211$.

**Passo 2 — verifichiamo che $211$ non è divisibile per $2,3,5,7$.** Come previsto dalla dimostrazione: $N\bmod p_i=1$ per ciascuno dei quattro primi (per costruzione, $N=p_1p_2p_3p_4+1$ ha resto $1$ nella divisione per ciascun $p_i$, essendo $p_1p_2p_3p_4$ multiplo di $p_i$).

**Passo 3 — troviamo il più piccolo fattore primo di $211$ per tentativi, fino a $\sqrt{211}\approx14{,}5$.** Proviamo $11,13$ (i primi dopo $7$ fino a $14$): se nessuno divide $211$, allora $211$ è esso stesso primo (per l'osservazione sul crivello: un composto $\le N$ ha sempre un fattore primo $\le\sqrt N$).

**Conclusione (verificare nella cella seguente):** in questo caso $211$ risulta esso stesso primo — un esempio di "nuovo" primo scovato dalla costruzione di Euclide, anche se, come osservato nella teoria, non è garantito che accada sempre (con $k=6$ l'esempio dato nella teoria produceva un numero composto).


```python
from sympy import isprime, factorint

primi_base = [2, 3, 5, 7]
N = 1
for p in primi_base:
    N *= p
N += 1
print(f"N = {'*'.join(map(str, primi_base))} + 1 = {N}")

for p in primi_base:
    print(f"  N mod {p} = {N % p}")

print(f"\n{N} è primo? {isprime(N)}")
print(f"Fattorizzazione di {N}: {factorint(N)}")

# Confrontiamo con l'esempio della teoria (k=6): 2*3*5*7*11*13 + 1
primi_base_6 = [2, 3, 5, 7, 11, 13]
N6 = 1
for p in primi_base_6:
    N6 *= p
N6 += 1
print(f"\nCon k=6: N = {N6}, primo? {isprime(N6)}, fattorizzazione: {factorint(N6)}")
```

## Riepilogo finale

- Un **numero primo** ha esattamente due divisori positivi, $1$ e se stesso; ogni altro intero $>1$ è composto e si fattorizza in primi (notebook 18.5).
- **Euclide**: i primi sono infiniti, dimostrato per assurdo considerando $N=p_1\cdots p_k+1$ e osservando che il suo fattore primo non può essere nella lista supposta completa.
- Il **crivello di Eratostene** trova tutti i primi fino a $N$ eliminando sistematicamente i multipli, fermandosi a $\sqrt N$.
- Il **teorema dei numeri primi** ($\pi(x)\sim x/\ln x$, enunciato senza dimostrazione) descrive quanto "diradano" i primi al crescere di $x$; congetture come Goldbach e i primi gemelli restano aperte.
