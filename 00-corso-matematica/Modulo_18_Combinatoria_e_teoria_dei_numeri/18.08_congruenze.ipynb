Ma restano due domande naturali che non abbiamo ancora affrontato: come risolvere **sistemi** di più congruenze simultanee (es. Questo notebook risponde a entrambe: il **teorema cinese del resto**, un risultato antico (Cina, III-V secolo) sorprendentemente utile ancora oggi, e il **piccolo teorema di Fermat**, che lega le potenze modulari alla primalità in un modo che sembra quasi magico la prima volta che lo si vede.


## Teoria, passo dopo passo

### Congruenze lineari

Una **congruenza lineare** è un'equazione $ax\equiv b\pmod n$, dove $a,b,n$ sono dati e cerchiamo tutte le classi $[x]\in\mathbb Z/n\mathbb Z$ che la soddisfano.

**Teorema.** Sia $d=\gcd(a,n)$. La congruenza $ax\equiv b\pmod n$ ha soluzione se e solo se $d\mid b$; in tal caso, ha esattamente $d$ soluzioni distinte modulo $n$.

**Dimostrazione (caso $d=1$, il cuore dell'argomento).** Se $\gcd(a,n)=1$, per il notebook 18.6 $[a]$ è invertibile in $\mathbb Z/n\mathbb Z$; moltiplicando entrambi i membri per $[a]^{-1}$ si ottiene l'**unica** soluzione $x\equiv a^{-1}b\pmod n$.

**Caso generale ($d=\gcd(a,n)>1$).** Se $d\nmid b$, non c'è soluzione: infatti $d\mid a$ e $d\mid n$, quindi se $ax\equiv b\pmod n$ per qualche $x$, cioè $ax-b=kn$ per qualche $k$, allora $d\mid(ax-kn)=b$ (chiusura per combinazioni lineari, notebook 18.5) — contraddizione. Se invece $d\mid b$, scriviamo $a=da'$, $b=db'$, $n=dn'$ con $\gcd(a',n')=1$ (proprietà del gcd, non dimostrata nel dettaglio qui): la congruenza equivale a $a'x\equiv b'\pmod{n'}$ (dividendo per $d$), che ha un'unica soluzione modulo $n'$ per il caso già dimostrato; questa singola classe modulo $n'$ corrisponde esattamente a $d$ classi distinte modulo $n$ (i rappresentanti $x_0,x_0+n',x_0+2n',\ldots,x_0+(d-1)n'$, tutti congrui modulo $n'$ ma distinti modulo $n$, essendo $n=dn'$). $\blacksquare$

### Il teorema cinese del resto

**Teorema.** Siano $n_1,\ldots,n_k$ interi positivi **coprimi a due a due** (cioè $\gcd(n_i,n_j)=1$ per $i\ne j$), e $a_1,\ldots,a_k$ interi qualsiasi. Il sistema
$$x\equiv a_1\pmod{n_1},\quad x\equiv a_2\pmod{n_2},\quad\ldots,\quad x\equiv a_k\pmod{n_k}$$
ha soluzione, ed è **unica modulo** $N=n_1n_2\cdots n_k$.

**Dimostrazione (costruttiva).** Per ogni $i$, poniamo $N_i=N/n_i$ (il prodotto di tutti gli $n_j$ tranne $n_i$). Poiché gli $n_j$ sono coprimi a due a due, $\gcd(N_i,n_i)=1$ (nessun fattore primo di $n_i$ divide alcun altro $n_j$, quindi nessuno divide $N_i$): per Bézout (notebook 18.5) esiste $M_i$ con $N_iM_i\equiv1\pmod{n_i}$.

Poniamo $x_0=\sum_{i=1}^ka_iN_iM_i$. Verifichiamo che $x_0$ soddisfa tutte le congruenze: fissato $j$, ogni termine $a_iN_iM_i$ con $i\ne j$ contiene il fattore $N_i$, che è multiplo di $n_j$ (per definizione, $N_i=N/n_i$ contiene tutti i fattori $n_l$ con $l\ne i$, incluso $n_j$ se $j\ne i$), quindi $a_iN_iM_i\equiv0\pmod{n_j}$ per $i\ne j$. Resta solo il termine $i=j$: $x_0\equiv a_jN_jM_j\equiv a_j\cdot1=a_j\pmod{n_j}$ (usando $N_jM_j\equiv1\pmod{n_j}$ per costruzione). Quindi $x_0$ soddisfa la $j$-esima congruenza, per ogni $j$: è una soluzione del sistema.

**Unicità modulo $N$.** Se $x_0,x_1$ sono entrambe soluzioni, allora $x_0\equiv x_1\pmod{n_i}$ per ogni $i$, cioè $n_i\mid(x_0-x_1)$ per ogni $i$. Essendo gli $n_i$ coprimi a due a due, il loro prodotto $N$ divide $(x_0-x_1)$ (fatto che generalizza, senza dimostrarlo nel dettaglio, l'osservazione che se due numeri coprimi dividono entrambi $m$, il loro prodotto divide $m$): quindi $x_0\equiv x_1\pmod N$. $\blacksquare$

### Il piccolo teorema di Fermat

**Teorema.** Se $p$ è primo e $p\nmid a$, allora $a^{p-1}\equiv1\pmod p$.

**Dimostrazione.** Consideriamo i $p-1$ numeri $a,2a,3a,\ldots,(p-1)a$, e riduciamoli modulo $p$. Affermiamo che sono, in qualche ordine, esattamente una permutazione di $1,2,\ldots,p-1$.

*Sono tutti non nulli modulo $p$:* se $ia\equiv0\pmod p$ per qualche $1\le i\le p-1$, allora $p\mid ia$; per il lemma di Euclide (notebook 18.5), $p\mid i$ oppure $p\mid a$. Ma $p\nmid a$ per ipotesi, e $p\nmid i$ perché $1\le i\le p-1<p$: contraddizione. Quindi ogni $ia\not\equiv0\pmod p$.

*Sono tutti distinti modulo $p$:* se $ia\equiv ja\pmod p$ per $1\le i,j\le p-1$, allora $p\mid(i-j)a$; di nuovo per il lemma di Euclide, $p\mid(i-j)$ oppure $p\mid a$; essendo $p\nmid a$, deve essere $p\mid(i-j)$, ma $|i-j|<p$ (entrambi in $[1,p-1]$), quindi $i-j=0$, cioè $i=j$.

Quindi $\{a\bmod p,2a\bmod p,\ldots,(p-1)a\bmod p\}=\{1,2,\ldots,p-1\}$ come insiemi (stessi $p-1$ elementi distinti, tutti non nulli, presi tra i $p-1$ residui non nulli possibili — quindi coincidono necessariamente). Moltiplicando tutti gli elementi di entrambi gli insiemi:
$$\prod_{i=1}^{p-1}(ia)\equiv\prod_{i=1}^{p-1}i\pmod p\quad\Longrightarrow\quad a^{p-1}\cdot(p-1)!\equiv(p-1)!\pmod p.$$
Poiché $p$ è primo e $p\nmid(p-1)!$ (nessuno dei fattori $1,\ldots,p-1$ è multiplo di $p$, per il lemma di Euclide applicato ripetutamente), $[(p-1)!]$ è invertibile modulo $p$ (notebook 18.6): moltiplicando entrambi i membri per il suo inverso,
$$a^{p-1}\equiv1\pmod p.\qquad\blacksquare$$

**Corollario.** Per ogni $a$ (anche se $p\mid a$), $a^p\equiv a\pmod p$ (se $p\mid a$ entrambi i membri sono $\equiv0$; altrimenti si moltiplica $a^{p-1}\equiv1$ per $a$).

### Applicazione: esponenziazione modulare rapida

Calcolare $a^k\bmod n$ per $k$ enorme non richiede $k$ moltiplicazioni: si usa la scrittura binaria di $k$, elevando ripetutamente al quadrato e riducendo modulo $n$ a ogni passo ("square and multiply"), in circa $\log_2k$ passi invece di $k$. Questa tecnica, insieme al piccolo teorema di Fermat, è alla base di molti algoritmi di crittografia moderna (che restano fuori dallo scopo di questo corso, ma la matematica che li rende possibili è esattamente quella vista qui).


```python
def crt(residui, moduli):
    # Risolve x = a_i mod n_i per liste di residui a_i e moduli n_i coprimi a due a due.
    N = 1
    for n in moduli:
        N *= n
    x0 = 0
    for a, n in zip(residui, moduli):
        Ni = N // n
        Mi = pow(Ni, -1, n)  # inverso di Ni modulo n
        x0 += a * Ni * Mi
    return x0 % N, N

# Sistema della teoria: x=2 mod 3, x=3 mod 5, x=2 mod 7
residui = [2, 3, 2]
moduli = [3, 5, 7]
x0, N = crt(residui, moduli)
print(f"Soluzione: x = {x0} mod {N}")
for a, n in zip(residui, moduli):
    print(f"  verifica: {x0} mod {n} = {x0 % n}  (atteso {a})")

# Verifica del piccolo teorema di Fermat per vari primi p e basi a
print()
for p in (5, 7, 11, 13):
    for a in range(1, p):
        assert pow(a, p - 1, p) == 1, f"Fermat fallito per a={a}, p={p}"
    print(f"Piccolo teorema di Fermat verificato per tutti gli a in [1,{p-1}], p={p}")
```

## Esercizio svolto e commentato

**Problema:** risolvi il sistema $x\equiv2\pmod3$, $x\equiv3\pmod5$, $x\equiv2\pmod7$ con il teorema cinese del resto, seguendo la costruzione della teoria passo per passo.

**Ragionamento:**

**Passo 1 — calcoliamo $N$ e gli $N_i$.** $N=3\cdot5\cdot7=105$. $N_1=105/3=35$, $N_2=105/5=21$, $N_3=105/7=15$.

**Passo 2 — troviamo $M_i$ tale che $N_iM_i\equiv1\pmod{n_i}$.**
- $N_1=35\equiv2\pmod3$ (essendo $35=33+2$): cerchiamo $M_1$ con $2M_1\equiv1\pmod3$; $M_1=2$ funziona ($2\cdot2=4\equiv1$).
- $N_2=21\equiv1\pmod5$ (essendo $21=20+1$): $M_2=1$ funziona banalmente ($1\cdot1\equiv1$).
- $N_3=15\equiv1\pmod7$ (essendo $15=14+1$): $M_3=1$ funziona banalmente.

**Passo 3 — combiniamo con la formula $x_0=\sum a_iN_iM_i$.**
$$x_0=2\cdot35\cdot2+3\cdot21\cdot1+2\cdot15\cdot1=140+63+30=233.$$

**Passo 4 — riduciamo modulo $N=105$.** $233=2\cdot105+23$, quindi $x_0\equiv23\pmod{105}$.

**Conclusione (verificare nella cella seguente):** $x=23$ dovrebbe soddisfare tutte e tre le congruenze: $23=3\cdot7+2\equiv2\pmod3$ ✓, $23=5\cdot4+3\equiv3\pmod5$ ✓, $23=7\cdot3+2\equiv2\pmod7$ ✓.


```python
x = 23
condizioni = [(2, 3), (3, 5), (2, 7)]
for a, n in condizioni:
    print(f"{x} mod {n} = {x % n}  (atteso {a}):  {'OK' if x % n == a else 'ERRORE'}")

# Confronto con la funzione crt() della cella precedente
def crt(residui, moduli):
    N = 1
    for n in moduli:
        N *= n
    x0 = 0
    for a, n in zip(residui, moduli):
        Ni = N // n
        Mi = pow(Ni, -1, n)
        x0 += a * Ni * Mi
    return x0 % N, N

x0, N = crt([2, 3, 2], [3, 5, 7])
print(f"\nFunzione crt(): x = {x0} mod {N}")
```

## Riepilogo finale

- Una **congruenza lineare** $ax\equiv b\pmod n$ ha soluzione sse $\gcd(a,n)\mid b$, con esattamente $\gcd(a,n)$ soluzioni distinte modulo $n$ quando ne ha.
- Il **teorema cinese del resto** garantisce che un sistema di congruenze con moduli coprimi a due a due abbia sempre soluzione unica modulo il prodotto dei moduli, con una costruzione esplicita basata su Bézout.
- Il **piccolo teorema di Fermat** ($a^{p-1}\equiv1\pmod p$ per $p$ primo, $p\nmid a$) si dimostra mostrando che $a,2a,\ldots,(p-1)a$ è una permutazione di $1,\ldots,p-1$ modulo $p$.
- L'**esponenziazione modulare rapida** (elevare al quadrato ripetutamente) rende praticabili calcoli con esponenti enormi, alla base della crittografia moderna.
