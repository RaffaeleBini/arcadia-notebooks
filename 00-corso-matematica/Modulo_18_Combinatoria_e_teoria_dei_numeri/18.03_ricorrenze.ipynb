# 18.3 — Ricorrenze

**Prerequisiti:** 18.2 Principio di inclusione-esclusione, 1.6 Equazioni di secondo grado, 6.17-6.19 Autovalori, autovettori, diagonalizzazione.


## Perché ci interessa

Nell'esercizio 3 del notebook 18.2 abbiamo trovato che i derangement soddisfano $D_n=(n-1)(D_{n-1}+D_{n-2})$: una regola che calcola $D_n$ a partire dai due termini precedenti, non una formula chiusa in $n$. Regole di questo tipo — dove ogni termine di una successione è definito in funzione dei termini precedenti — si chiamano **ricorrenze**, e sono ovunque in matematica discreta: contare configurazioni spesso porta naturalmente a una regola ricorsiva ("il numero di modi con $n$ oggetti dipende dai modi con $n-1$ o $n-2$ oggetti"), prima ancora che a una formula esplicita.

Una ricorrenza è comodissima per calcolare (basta iterare), ma nasconde il comportamento a lungo termine: cresce come $n$? come $2^n$? Il problema di questo notebook è passare da una **regola ricorsiva** a una **formula chiusa**, usando un metodo generale che funziona per un'intera classe di ricorrenze — non solo caso per caso.


## Teoria, passo dopo passo

### L'esempio guida: la successione di Fibonacci

La successione più famosa definita per ricorrenza è quella di Fibonacci:
$$F_0=0,\quad F_1=1,\quad F_n=F_{n-1}+F_{n-2}\quad(n\ge2).$$
Nasce da un problema di conteggio (il classico problema dei conigli di Fibonacci, o — più utile per noi — contare le stringhe binarie di lunghezza $n$ senza due $1$ consecutivi, o le sequenze $\pm1$ come nella passeggiata aleatoria del notebook 17.10 con vincoli). È una **ricorrenza lineare a coefficienti costanti di ordine $2$**: "lineare" perché $F_n$ è combinazione lineare dei termini precedenti (nessun quadrato, nessun prodotto $F_{n-1}F_{n-2}$), "a coefficienti costanti" perché i coefficienti ($1$ e $1$) non dipendono da $n$, "di ordine $2$" perché servono i due termini precedenti.

### Il metodo dell'equazione caratteristica

**Idea.** Cerchiamo soluzioni della forma $a_n=r^n$ per qualche $r\ne0$ (una scelta motivata: le uniche successioni che "si autoriproducono" per moltiplicazione, come vedremo, sono le potenze geometriche). Sostituendo $a_n=r^n$ nella ricorrenza $a_n=a_{n-1}+a_{n-2}$:
$$r^n=r^{n-1}+r^{n-2}\quad\Longleftrightarrow\quad r^2=r+1\quad\Longleftrightarrow\quad r^2-r-1=0$$
(dividendo per $r^{n-2}\ne0$). Questa è l'**equazione caratteristica** della ricorrenza. Le sue radici (formula per equazioni di secondo grado, notebook 1.6) sono
$$\varphi=\frac{1+\sqrt5}2\approx1{,}618,\qquad\psi=\frac{1-\sqrt5}2\approx-0{,}618$$
($\varphi$ è la famosa **sezione aurea**).

**Perché la soluzione generale è una combinazione lineare delle due potenze.** L'insieme delle successioni $(a_n)$ che soddisfano $a_n=a_{n-1}+a_{n-2}$ per ogni $n\ge2$ (senza fissare $a_0,a_1$) è uno **spazio vettoriale** nel senso del notebook 6.3: se $(a_n)$ e $(b_n)$ soddisfano la ricorrenza, anche $(\alpha a_n+\beta b_n)$ la soddisfa, per linearità della relazione. Inoltre questo spazio ha dimensione $2$: una successione che soddisfa la ricorrenza è determinata univocamente dai suoi primi due valori $a_0,a_1$ (tutto il resto si calcola per iterazione), quindi la mappa "successione $\mapsto(a_0,a_1)$" è un isomorfismo (notebook 6.13) tra lo spazio delle soluzioni e $\mathbb R^2$. Se $\varphi\ne\psi$ (radici distinte), le due successioni $(\varphi^n)$ e $(\psi^n)$ sono linearmente indipendenti (i vettori $(\varphi^0,\varphi^1)=(1,\varphi)$ e $(\psi^0,\psi^1)=(1,\psi)$ non sono proporzionali, perché $\varphi\ne\psi$), quindi formano una **base** (notebook 6.8) dello spazio delle soluzioni: ogni soluzione si scrive in modo unico come
$$a_n=A\varphi^n+B\psi^n.$$

**Determiniamo $A,B$ dalle condizioni iniziali di Fibonacci.** Da $F_0=0=A+B$ e $F_1=1=A\varphi+B\psi$: dalla prima, $B=-A$; sostituendo, $A(\varphi-\psi)=1$, e $\varphi-\psi=\sqrt5$, quindi $A=1/\sqrt5$, $B=-1/\sqrt5$. Si ottiene la **formula di Binet**:
$$F_n=\frac{\varphi^n-\psi^n}{\sqrt5}=\frac1{\sqrt5}\left[\left(\frac{1+\sqrt5}2\right)^n-\left(\frac{1-\sqrt5}2\right)^n\right].$$
Sorprendente: una formula con radici quadrate produce sempre un **intero** per ogni $n$ — i due termini irrazionali si compensano esattamente.

**Il comportamento asintotico.** Poiché $|\psi|<1$, $\psi^n\to0$: quindi $F_n\sim\varphi^n/\sqrt5$ per $n$ grande — crescita **esponenziale** di rapporto $\varphi$, la sezione aurea. Infatti $F_n/F_{n-1}\to\varphi$: è un fatto classico, e lo verificheremo nella cella di codice.

### Il caso di radici coincidenti (cenno)

Se l'equazione caratteristica ha una radice doppia $r$ (discriminante nullo, notebook 1.6), la soluzione generale non è più $Ar^n+Br^n$ (che collasserebbe a una sola costante libera, insufficiente per uno spazio di dimensione $2$), ma
$$a_n=(A+Bn)r^n$$
(si può verificare per sostituzione diretta che anche $nr^n$ risolve la ricorrenza quando $r$ è radice doppia). Non approfondiamo oltre questo cenno, che esula dagli esempi di questo notebook.

### Un secondo metodo: la forma matriciale

La ricorrenza di Fibonacci si può riscrivere come un'equazione tra vettori:
$$\begin{pmatrix}F_{n+1}\\F_n\end{pmatrix}=\underbrace{\begin{pmatrix}1&1\\1&0\end{pmatrix}}_{=:M}\begin{pmatrix}F_n\\F_{n-1}\end{pmatrix}\quad\Longrightarrow\quad\begin{pmatrix}F_{n+1}\\F_n\end{pmatrix}=M^n\begin{pmatrix}F_1\\F_0\end{pmatrix}.$$
Per calcolare $M^n$ in modo efficiente, **diagonalizziamo** $M$ (notebook 6.19): gli autovalori di $M$ sono le radici del suo polinomio caratteristico $\det(M-\lambda I)=\lambda^2-\lambda-1=0$ — **la stessa equazione caratteristica di prima**, non a caso: $\varphi,\psi$. Scrivendo $M=PDP^{-1}$ con $D=\mathrm{diag}(\varphi,\psi)$, si ha $M^n=PD^nP^{-1}=P\,\mathrm{diag}(\varphi^n,\psi^n)\,P^{-1}$, e portando a termine il calcolo si ritrova esattamente la formula di Binet. Questo secondo metodo non è solo un'alternativa: mostra che **il metodo dell'equazione caratteristica è, in sostanza, la diagonalizzazione della matrice di transizione della ricorrenza** — lo stesso fenomeno algebrico visto da due angolazioni diverse, un collegamento diretto col Modulo 6.


```python
import numpy as np

# Verifichiamo la formula di Binet confrontandola con Fibonacci calcolato per iterazione,
# e verifichiamo che F_n/F_(n-1) -> phi (sezione aurea).

def fibonacci_iterativo(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

phi = (1 + np.sqrt(5)) / 2
psi = (1 - np.sqrt(5)) / 2

print(f"phi (sezione aurea) = {phi:.6f}")
print(f"{'n':>3} {'F_n iterativo':>15} {'Binet':>15} {'F_n/F_(n-1)':>15}")
prec = None
for n in range(1, 16):
    it = fibonacci_iterativo(n)
    binet = round((phi**n - psi**n) / np.sqrt(5))
    rapporto = it / prec if prec else float('nan')
    print(f"{n:>3} {it:>15} {binet:>15} {rapporto:>15.6f}")
    prec = it

# Verifica via la potenza della matrice di transizione (metodo matriciale)
M = np.array([[1, 1], [1, 0]])
n = 15
M_n = np.linalg.matrix_power(M, n)
print(f"\nM^{n} @ (F_1, F_0) = {M_n @ np.array([1, 0])}  (atteso: (F_{n+1}, F_{n}) = ({fibonacci_iterativo(n+1)}, {fibonacci_iterativo(n)}))")
```

## Esercizio svolto e commentato

**Problema:** risolvi per esteso la ricorrenza $a_n=5a_{n-1}-6a_{n-2}$, con $a_0=1,a_1=0$.

**Ragionamento:**

**Passo 1 — scriviamo l'equazione caratteristica.** Cerchiamo $a_n=r^n$: sostituendo, $r^n=5r^{n-1}-6r^{n-2}$, e dividendo per $r^{n-2}$: $r^2=5r-6$, cioè $r^2-5r+6=0$.

**Passo 2 — risolviamo l'equazione di secondo grado (notebook 1.6).** $r=\dfrac{5\pm\sqrt{25-24}}2=\dfrac{5\pm1}2$, quindi $r_1=3,r_2=2$: radici reali e distinte.

**Passo 3 — scriviamo la soluzione generale.** Essendo le radici distinte, per la teoria vista sopra $a_n=A\cdot3^n+B\cdot2^n$.

**Passo 4 — usiamo le condizioni iniziali per trovare $A,B$.** Da $a_0=1$: $A+B=1$. Da $a_1=0$: $3A+2B=0$. Dalla prima, $B=1-A$; sostituendo nella seconda: $3A+2(1-A)=0\Rightarrow A+2=0\Rightarrow A=-2$, quindi $B=3$.

**Conclusione:** $a_n=-2\cdot3^n+3\cdot2^n$ (da verificare nella cella seguente confrontando con il calcolo iterativo diretto).

**Perché il metodo funziona sempre in questo caso.** Ogni volta che l'equazione caratteristica ha due radici reali distinte, i passi 1-4 si ripetono identici: scrivere l'equazione caratteristica, risolverla, imporre le condizioni iniziali su un sistema lineare $2\times2$ nei coefficienti $A,B$ (sempre risolubile in modo unico quando le radici sono distinte, essendo il determinante del sistema $r_2-r_1\ne0$).


```python
def a_iterativo(n):
    a0, a1 = 1, 0
    if n == 0:
        return a0
    if n == 1:
        return a1
    a, b = a0, a1
    for _ in range(2, n + 1):
        a, b = b, 5 * b - 6 * a
    return b

def a_formula(n):
    return -2 * 3**n + 3 * 2**n

print(f"{'n':>3} {'iterativo':>12} {'formula':>12}")
for n in range(8):
    print(f"{n:>3} {a_iterativo(n):>12} {a_formula(n):>12}")
```

## Riepilogo finale

- Una **ricorrenza lineare a coefficienti costanti** definisce ogni termine come combinazione lineare fissa dei precedenti; il metodo dell'**equazione caratteristica** (sostituire $a_n=r^n$) la trasforma in un'equazione polinomiale.
- Se le radici $r_1,r_2$ sono distinte, la soluzione generale è $a_n=Ar_1^n+Br_2^n$, perché lo spazio delle soluzioni ha dimensione $2$ (notebook 6.3, 6.8) e $\{r_1^n\},\{r_2^n\}$ ne sono una base; le condizioni iniziali fissano $A,B$ in modo unico.
- Fibonacci: $F_n=(\varphi^n-\psi^n)/\sqrt5$ (formula di Binet), con crescita asintotica $F_n\sim\varphi^n/\sqrt5$ governata dalla sezione aurea $\varphi$.
- Il metodo dell'equazione caratteristica **è** la diagonalizzazione (Modulo 6) della matrice di transizione della ricorrenza: due nomi per lo stesso fenomeno algebrico.
- Il caso di radice doppia richiede la forma $a_n=(A+Bn)r^n$; ricorrenze non omogenee (col termine noto, come Torri di Hanoi) richiedono un termine aggiuntivo, spesso trovabile per tentativi.
