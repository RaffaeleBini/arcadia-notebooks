**Prerequisiti:** 18.8 Congruenze, 18.6 Aritmetica modulare, 18.5 Teoria dei numeri elementare.


## Perché ci interessa

Il piccolo teorema di Fermat (notebook 18.8) parla solo di moduli **primi**. Ma cosa succede modulo un $n$ composto, come $10$ o $360$? Serve un ingrediente che "conti" quanti sono gli elementi invertibili di $\mathbb Z/n\mathbb Z$ — esattamente il ruolo che, per $n=p$ primo, era giocato dal numero $p-1$ nella dimostrazione di Fermat.

Questo notebook introduce le **funzioni aritmetiche**: funzioni definite sugli interi positivi che dipendono dalla loro struttura di divisori. La protagonista è la **funzione di Eulero** $\varphi(n)$, che generalizza Fermat a ogni modulo; ma studieremo anche altre funzioni aritmetiche classiche (numero e somma dei divisori, funzione di Möbius), tutte legate da una proprietà comune — la **moltiplicatività** — che le rende calcolabili a partire dalla sola fattorizzazione in primi.


## Teoria, passo dopo passo

### La funzione di Eulero

La **funzione di Eulero** $\varphi(n)$ conta gli interi in $\{1,\ldots,n\}$ coprimi con $n$. Per il notebook 18.6, questo è esattamente il numero di elementi **invertibili** di $\mathbb Z/n\mathbb Z$ — l'insieme $(\mathbb Z/n\mathbb Z)^*$, che forma un gruppo moltiplicativo (notebook 11.1) di ordine $\varphi(n)$.

**Caso $n=p$ primo:** ogni intero da $1$ a $p-1$ è coprimo con $p$ (essendo $p$ primo), quindi $\varphi(p)=p-1$ — coerente col numero di elementi coinvolti nella dimostrazione di Fermat.

**Caso $n=p^k$ (potenza di primo):** tra $1$ e $p^k$, i **non** coprimi con $p^k$ sono esattamente i multipli di $p$ (essendo $p$ l'unico fattore primo di $p^k$): ce ne sono $p^k/p=p^{k-1}$. Quindi
$$\varphi(p^k)=p^k-p^{k-1}=p^k\Big(1-\frac1p\Big).$$

### Il teorema di Eulero-Fermat

**Teorema.** Se $\gcd(a,n)=1$, allora $a^{\varphi(n)}\equiv1\pmod n$.

**Dimostrazione.** Identica nella struttura a quella di Fermat (notebook 18.8), sostituendo $\{1,\ldots,p-1\}$ con l'insieme $R=\{r_1,\ldots,r_{\varphi(n)}\}$ dei residui invertibili modulo $n$ (cioè coprimi con $n$). Si mostra che $\{ar_1,\ldots,ar_{\varphi(n)}\}$, ridotti modulo $n$, sono ancora una permutazione di $R$: ciascun $ar_i$ resta coprimo con $n$ (prodotto di due elementi coprimi con $n$ è coprimo con $n$, un fatto che segue da Bézout: se $\gcd(a,n)=\gcd(r_i,n)=1$, esistono $x,y,x',y'$ con $ax+ny=1$, $r_ix'+ny'=1$; moltiplicando, $ar_i(xx')+n(\ldots)=1$, quindi $\gcd(ar_i,n)=1$), e sono distinti modulo $n$ per lo stesso argomento cancellativo usato in Fermat. Moltiplicando tutti gli elementi di $R$ e confrontando con il prodotto di $\{ar_i\}$, si ottiene $a^{\varphi(n)}\prod r_i\equiv\prod r_i\pmod n$, e semplificando (essendo $\prod r_i$ invertibile, prodotto di invertibili) si ottiene la tesi. $\blacksquare$

Per $n=p$ primo, $\varphi(p)=p-1$: si ritrova esattamente il piccolo teorema di Fermat come **caso particolare**.

### Funzioni moltiplicative

Una funzione aritmetica $f:\mathbb Z_{>0}\to\mathbb R$ si dice **moltiplicativa** se $f(mn)=f(m)f(n)$ ogni volta che $\gcd(m,n)=1$ (attenzione: non per **ogni** coppia $m,n$, solo per quelle coprime).

**Teorema.** $\varphi$ è moltiplicativa.

**Idea della dimostrazione (senza tutti i dettagli tecnici).** Se $\gcd(m,n)=1$, il teorema cinese del resto (notebook 18.8) garantisce una corrispondenza biunivoca tra $\mathbb Z/mn\mathbb Z$ e le coppie $(\mathbb Z/m\mathbb Z)\times(\mathbb Z/n\mathbb Z)$ (a $x\bmod mn$ associamo $(x\bmod m,x\bmod n)$); si può verificare che questa corrispondenza preserva l'invertibilità (un elemento è invertibile modulo $mn$ sse è invertibile sia modulo $m$ sia modulo $n$, essendo $\gcd(x,mn)=1\iff\gcd(x,m)=1$ e $\gcd(x,n)=1$), quindi conta gli elementi invertibili nello stesso modo su entrambi i lati: $\varphi(mn)=\varphi(m)\varphi(n)$.

**Conseguenza — formula esplicita.** Se $n=p_1^{\alpha_1}\cdots p_r^{\alpha_r}$ è la fattorizzazione in primi (notebook 18.5), essendo le potenze $p_i^{\alpha_i}$ coprime a due a due, la moltiplicatività dà
$$\varphi(n)=\prod_{i=1}^r\varphi(p_i^{\alpha_i})=\prod_{i=1}^rp_i^{\alpha_i}\Big(1-\frac1{p_i}\Big)=n\prod_{p\mid n}\Big(1-\frac1p\Big),$$
dove il prodotto è sui primi $p$ che dividono $n$ (senza ripetizione). Questa formula rende $\varphi(n)$ calcolabile immediatamente dalla fattorizzazione.

### Altre funzioni aritmetiche moltiplicative

- $\tau(n)$ (o $d(n)$): il **numero di divisori** positivi di $n$. Se $n=\prod p_i^{\alpha_i}$, ogni divisore si ottiene scegliendo, per ciascun $p_i$, un esponente tra $0$ e $\alpha_i$ ($\alpha_i+1$ scelte indipendenti, principio moltiplicativo, notebook 18.1): quindi $\tau(n)=\prod_i(\alpha_i+1)$. Moltiplicativa (verifica diretta dalla formula).
- $\sigma(n)$: la **somma dei divisori** positivi di $n$. Per $n=p^\alpha$, $\sigma(p^\alpha)=1+p+\cdots+p^\alpha=\dfrac{p^{\alpha+1}-1}{p-1}$ (somma geometrica, notebook 4.8); moltiplicativa, quindi $\sigma(n)=\prod_i\dfrac{p_i^{\alpha_i+1}-1}{p_i-1}$.

### La funzione di Möbius e la formula di inversione

La **funzione di Möbius** è definita da
$$\mu(n)=\begin{cases}1&n=1\\(-1)^r&n=p_1p_2\cdots p_r\text{, primi distinti}\\0&\text{se }n\text{ ha un fattore primo ripetuto (non libero da quadrati)}\end{cases}$$
È moltiplicativa (verifica diretta dalla definizione, distinguendo i casi).

**Identità fondamentale.** $\displaystyle\sum_{d\mid n}\mu(d)=\begin{cases}1&n=1\\0&n>1\end{cases}$.

**Dimostrazione (per $n>1$).** Sia $n=p_1^{\alpha_1}\cdots p_r^{\alpha_r}$. I divisori $d$ con $\mu(d)\ne0$ sono esattamente i prodotti di sottoinsiemi distinti di $\{p_1,\ldots,p_r\}$ (i divisori "liberi da quadrati" di $n$); per un sottoinsieme di taglia $k$, $\mu(d)=(-1)^k$. Raggruppando per taglia $k$ (ci sono $\binom rk$ sottoinsiemi di taglia $k$, notebook 18.1):
$$\sum_{d\mid n}\mu(d)=\sum_{k=0}^r\binom rk(-1)^k=(1-1)^r=0$$
per il binomio di Newton (notebook 18.1), essendo $r\ge1$ (perché $n>1$ ha almeno un fattore primo). $\blacksquare$

**Formula di inversione di Möbius.** Se $g(n)=\sum_{d\mid n}f(d)$ per ogni $n$, allora $f(n)=\sum_{d\mid n}\mu(d)g(n/d)$ (enunciata senza dimostrazione completa, che è un calcolo diretto — uno scambio di ordine di somma, usando l'identità appena dimostrata — ma tecnicamente lungo da scrivere per esteso). Questa formula "inverte" una somma sui divisori, ed è lo strumento con cui si dimostrano molte identità della teoria dei numeri, incluso un fatto notevole:
$$\sum_{d\mid n}\varphi(d)=n$$
(intuizione: ogni frazione $i/n$ con $1\le i\le n$, ridotta ai minimi termini, ha un denominatore $d$ che divide $n$; per ciascun divisore $d$ di $n$, ci sono esattamente $\varphi(d)$ frazioni con quel denominatore ridotto — quindi sommando su tutti i divisori $d$ di $n$ si contano tutte le $n$ frazioni $1/n,2/n,\ldots,n/n$ esattamente una volta).


```python
from sympy import totient, divisor_count, divisor_sigma, factorint, mobius, divisors
from math import gcd

# Verifica della moltiplicativita' di phi, tau, sigma su coppie coprime,
# e della formula esplicita di phi dalla fattorizzazione.

def phi_da_fattorizzazione(n):
    fatt = factorint(n)
    risultato = n
    for p in fatt:
        risultato = risultato * (p - 1) // p
    return risultato

for n in (12, 30, 100, 360):
    print(f"n={n}: phi(n)={totient(n)} (formula: {phi_da_fattorizzazione(n)}), "
          f"tau(n)={divisor_count(n)}, sigma(n)={divisor_sigma(n)}, fattorizzazione={factorint(n)}")

print()
m, n = 8, 9  # coprimi
print(f"Verifica moltiplicativita' (m={m}, n={n}, gcd={gcd(m, n)}):")
print(f"  phi({m*n}) = {totient(m*n)},  phi({m})*phi({n}) = {totient(m)*totient(n)}")
print(f"  tau({m*n}) = {divisor_count(m*n)},  tau({m})*tau({n}) = {divisor_count(m)*divisor_count(n)}")

print()
n = 360
somma = sum(totient(d) for d in divisors(n))
print(f"Verifica sum_{{d|n}} phi(d) = n per n={n}: somma = {somma}, n = {n}")

print()
n = 30
somma_mu = sum(mobius(d) for d in divisors(n))
print(f"Verifica sum_{{d|n}} mu(d) = 0 per n={n} (n>1): somma = {somma_mu}")
```

## Esercizio svolto e commentato

**Problema:** calcola $\varphi(360)$ dalla fattorizzazione, e verifica numericamente il teorema di Eulero-Fermat per $a=7$, $n=360$.

**Ragionamento:**

**Passo 1 — fattorizziamo $360$.** $360=2^3\cdot3^2\cdot5$.

**Passo 2 — applichiamo la formula $\varphi(n)=n\prod_{p\mid n}(1-1/p)$.**
$$\varphi(360)=360\cdot\Big(1-\frac12\Big)\Big(1-\frac13\Big)\Big(1-\frac15\Big)=360\cdot\frac12\cdot\frac23\cdot\frac45.$$
Calcoliamo passo passo: $360\cdot\frac12=180$; $180\cdot\frac23=120$; $120\cdot\frac45=96$. Quindi $\varphi(360)=96$.

**Passo 3 — verifichiamo $\gcd(7,360)=1$.** $7$ è primo e non divide $360=2^3\cdot3^2\cdot5$ (nessuno dei fattori primi di $360$ è $7$): coprimi, quindi il teorema di Eulero-Fermat si applica.

**Passo 4 — prevediamo $7^{96}\equiv1\pmod{360}$**, da verificare nella cella seguente con l'esponenziazione modulare rapida (notebook 18.8).

**Perché non possiamo usare direttamente Fermat "semplice".** $360$ non è primo: il teorema del notebook 18.8 non si applica direttamente. Serve la sua generalizzazione, il teorema di Eulero-Fermat, con $\varphi(360)$ al posto di "$p-1$" — esattamente il motivo per cui $\varphi$ è stata introdotta.


```python
from sympy import totient, factorint

n = 360
a = 7
phi_n = int(totient(n))

print(f"Fattorizzazione di {n}: {factorint(n)}")
print(f"phi({n}) = {phi_n}")
print(f"gcd({a},{n}) = 1 ? {__import__('math').gcd(a, n) == 1}")
print(f"{a}^{phi_n} mod {n} = {pow(a, phi_n, n)}  (atteso: 1, per Eulero-Fermat)")
```

## Riepilogo finale

- La **funzione di Eulero** $\varphi(n)$ conta gli invertibili modulo $n$; il **teorema di Eulero-Fermat** ($a^{\varphi(n)}\equiv1\pmod n$ per $\gcd(a,n)=1$) generalizza Fermat a moduli composti.
- Una funzione è **moltiplicativa** se $f(mn)=f(m)f(n)$ per $m,n$ coprimi: questo la rende calcolabile dalla sola fattorizzazione in primi, tramite il teorema cinese del resto (notebook 18.8).
- $\varphi(n)=n\prod_{p\mid n}(1-1/p)$; $\tau(n)=\prod_i(\alpha_i+1)$ (numero di divisori); $\sigma(n)$ (somma dei divisori) tramite somme geometriche.
- La **funzione di Möbius** $\mu(n)$ soddisfa $\sum_{d\mid n}\mu(d)=[n=1]$, e genera la **formula di inversione di Möbius**, uno strumento per "invertire" somme sui divisori — usato per dimostrare identità come $\sum_{d\mid n}\varphi(d)=n$.
