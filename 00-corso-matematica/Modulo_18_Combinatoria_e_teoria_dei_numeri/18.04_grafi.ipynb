## Teoria, passo dopo passo

### Definizioni di base

Un **grafo (semplice, non orientato)** è una coppia $G=(V,E)$ dove $V$ è un insieme finito di **vertici** (o nodi) ed $E$ è un insieme di coppie non ordinate $\{u,v\}$ con $u,v\in V$, $u\ne v$, dette **archi** (o spigoli, o lati): "semplice" perché non ammettiamo archi ripetuti né archi da un vertice a se stesso (**cappi**), "non orientato" perché l'arco $\{u,v\}$ non ha una direzione preferita.

Il **grado** $\deg(v)$ di un vertice $v$ è il numero di archi che lo toccano, cioè $|\{e\in E:v\in e\}|$.

Un **cammino** da $u$ a $v$ è una sequenza di vertici $u=v_0,v_1,\ldots,v_k=v$ tale che $\{v_{i-1},v_i\}\in E$ per ogni $i$; un **ciclo** è un cammino chiuso ($v_0=v_k$) che non ripete vertici (a parte l'estremo iniziale/finale). Il grafo è **connesso** se esiste un cammino tra ogni coppia di vertici.

### Il lemma delle strette di mano

**Teorema.** In ogni grafo $G=(V,E)$, $\sum_{v\in V}\deg(v)=2|E|$.

**Dimostrazione (per doppio conteggio).** Contiamo le coppie $(v,e)$ con $v\in V$, $e\in E$, $v\in e$ (cioè "il vertice $v$ è un estremo dell'arco $e$"), in due modi diversi.
- **Per vertice:** fissato $v$, il numero di archi che lo contengono è per definizione $\deg(v)$; sommando su tutti i vertici, il totale è $\sum_{v}\deg(v)$.
- **Per arco:** fissato $e=\{u,v\}$, esso contribuisce a esattamente $2$ coppie (una con $v=u$, una con $v=v$, i due estremi); sommando su tutti gli archi, il totale è $2|E|$.

Contando lo stesso insieme di coppie in due modi si ottiene lo stesso numero: $\sum_v\deg(v)=2|E|$. $\blacksquare$

**Corollario immediato:** il numero di vertici di grado dispari è sempre **pari** (altrimenti la somma dei gradi, che deve essere pari per il teorema, sarebbe dispari — una somma con un numero dispari di addendi dispari è dispari).

### Alberi

Un **albero** è un grafo connesso e **aciclico** (senza cicli). Gli alberi sono la struttura "minimale" per collegare $n$ vertici.

**Teorema.** Un albero con $n$ vertici ha esattamente $n-1$ archi.

**Dimostrazione (per induzione forte su $n$, notebook 10.4).**

*Base $n=1$:* un albero con un solo vertice non ha archi (nessun cappio ammesso): $n-1=0$ archi. Vero.

*Passo induttivo.* Supponiamo la tesi vera per ogni albero con meno di $n$ vertici, e sia $T$ un albero con $n\ge2$ vertici. Un albero con almeno $2$ vertici ha sempre una **foglia** (un vertice di grado $1$) — fatto che usiamo senza dimostrare nel dettaglio (segue dall'assenza di cicli: seguendo un cammino che non ripete vertici, essendo $V$ finito, deve terminare in un vertice senza "via d'uscita" nuova, cioè di grado $1$). Rimuoviamo una foglia $v$ e l'unico arco che la tocca: il grafo risultante $T'$ ha $n-1$ vertici, resta connesso (nessun cammino tra gli altri vertici passava per $v$, essendo $v$ di grado $1$) e resta aciclico (rimuovere vertici e archi non può creare cicli), quindi $T'$ è ancora un albero. Per ipotesi induttiva, $T'$ ha $(n-1)-1=n-2$ archi. Rimettendo l'arco tolto, $T$ ha $n-2+1=n-1$ archi. $\blacksquare$

Questo teorema è coerente con l'intuizione: un albero è "il minimo indispensabile" per connettere $n$ vertici — un solo arco in meno ($n-2$) sconnetterebbe qualcosa, un solo arco in più ($n$) creerebbe necessariamente un ciclo (fatti che non dimostriamo qui, ma che rendono gli alberi la struttura connessa "più economica possibile").

### La matrice di adiacenza

Numerando i vertici $v_1,\ldots,v_n$, la **matrice di adiacenza** $A\in\{0,1\}^{n\times n}$ (notebook 6.10) ha $A_{ij}=1$ se $\{v_i,v_j\}\in E$, altrimenti $0$ (e $A_{ii}=0$, nessun cappio). È simmetrica, perché il grafo non è orientato.

**Teorema.** Per ogni $k\ge1$, $(A^k)_{ij}$ è il numero di cammini di lunghezza esattamente $k$ da $v_i$ a $v_j$ (dove un "cammino" qui può ripassare per vertici già visitati — una passeggiata, più precisamente).

**Dimostrazione (per induzione su $k$).**

*Base $k=1$:* $(A^1)_{ij}=A_{ij}$ vale $1$ se c'è un arco (cammino di lunghezza $1$) tra $v_i,v_j$, $0$ altrimenti: coincide col numero di cammini di lunghezza $1$. Vero.

*Passo induttivo.* Supponiamo che $(A^k)_{ij}$ conti i cammini di lunghezza $k$ da $v_i$ a $v_j$. Per definizione di prodotto tra matrici (notebook 6.10),
$$(A^{k+1})_{ij}=(A^k A)_{ij}=\sum_{m=1}^n(A^k)_{im}A_{mj}.$$
Ogni cammino di lunghezza $k+1$ da $v_i$ a $v_j$ è, in modo univoco, un cammino di lunghezza $k$ da $v_i$ a un qualche vertice intermedio $v_m$ (contato da $(A^k)_{im}$, per ipotesi induttiva) seguito da un arco $v_m\to v_j$ (che esiste sse $A_{mj}=1$): sommando su tutte le scelte del vertice intermedio $v_m$ si contano esattamente, e senza sovraconta, tutti i cammini di lunghezza $k+1$ (principio additivo, notebook 18.1, applicato ai casi disgiunti "quale vertice intermedio $v_m$"). $\blacksquare$

Questo teorema è un ottimo esempio di come una nozione puramente algebrica (potenza di matrice, Modulo 6) codifichi esattamente un conteggio combinatorio.

### Cammini euleriani: il problema dei ponti di Königsberg

Il problema storico che ha dato inizio alla teoria dei grafi (Eulero, 1736): la città di Königsberg aveva $4$ zone di terra collegate da $7$ ponti; era possibile attraversare ogni ponte **esattamente una volta** e tornare al punto di partenza? Modellando le $4$ zone come vertici e i $7$ ponti come archi (con archi ripetuti tra le stesse zone: un **multigrafo**, generalizzazione di un grafo semplice), la domanda diventa: esiste un **circuito euleriano** (un cammino chiuso che percorre ogni arco esattamente una volta)?

**Teorema (Eulero).** Un grafo connesso ammette un circuito euleriano se e solo se ogni vertice ha grado pari.

**Dimostrazione della condizione necessaria (l'unica che dimostriamo per esteso).** Supponiamo che esista un circuito euleriano. Ogni volta che il circuito passa per un vertice $v$ (diverso dal punto di partenza/arrivo), "consuma" due archi incidenti a $v$: uno in entrata, uno in uscita. Poiché il circuito percorre **ogni** arco esattamente una volta ed è chiuso, ogni visita a $v$ consuma esattamente $2$ dei suoi archi incidenti, quindi $\deg(v)$ dev'essere un multiplo di $2$, cioè pari (lo stesso vale per il vertice di partenza/arrivo, contando anche la coppia "arco di partenza, arco di arrivo finale"). $\blacksquare$

*(La condizione sufficiente, cioè che il grado pari ovunque basti a garantire l'esistenza di un simile circuito, richiede una costruzione esplicita — tipicamente per induzione sul numero di archi, decomponendo il grafo in cicli — che non riportiamo qui.)*

A Königsberg, i $4$ vertici avevano gradi $3,3,3,5$ — tutti dispari: per il teorema (nella direzione dimostrata), **nessun** circuito euleriano può esistere. È il primo esempio nella storia della matematica di un risultato di **non esistenza** dimostrato con un argomento puramente combinatorio, senza dover tentare tutti i percorsi possibili.


```python
import numpy as np

# Rappresentiamo un piccolo grafo (il grafo completo K4, per semplicita') tramite la sua
# matrice di adiacenza, e verifichiamo il lemma delle strette di mano e il teorema sui cammini.

vertici = ["A", "B", "C", "D"]
n = len(vertici)
archi = [("A", "B"), ("A", "C"), ("A", "D"), ("B", "C"), ("B", "D"), ("C", "D")]  # K4: tutti connessi

idx = {v: i for i, v in enumerate(vertici)}
A = np.zeros((n, n), dtype=int)
for u, v in archi:
    A[idx[u], idx[v]] = 1
    A[idx[v], idx[u]] = 1

print("Matrice di adiacenza:")
print(A)

gradi = A.sum(axis=1)
print(f"\nGradi dei vertici: {dict(zip(vertici, gradi))}")
print(f"Somma dei gradi: {gradi.sum()}, 2*|E| = {2 * len(archi)}  (lemma delle strette di mano)")

# Numero di cammini di lunghezza 2 da A a C, letto da A^2
A2 = A @ A
print(f"\n(A^2)[A,C] = {A2[idx['A'], idx['C']]}  = numero di cammini di lunghezza 2 da A a C")
# Verifica manuale: A-B-C e A-D-C sono gli unici due cammini di lunghezza 2 da A a C in K4
```

## Esercizio svolto e commentato

**Problema:** verifica che il grafo "a stella" con un vertice centrale $C$ collegato a $4$ foglie $F_1,F_2,F_3,F_4$ (e nessun altro arco) è un albero, e verifica il teorema "$n$ vertici $\Rightarrow n-1$ archi" per questo esempio.

**Ragionamento:**

**Passo 1 — verifichiamo che è connesso.** Ogni foglia è collegata direttamente a $C$ (cammino di lunghezza $1$); due foglie qualsiasi $F_i,F_j$ sono collegate dal cammino $F_i,C,F_j$ (lunghezza $2$). Ogni coppia di vertici ha un cammino che li collega: connesso.

**Passo 2 — verifichiamo che è aciclico.** Un ciclo dovrebbe tornare al punto di partenza senza ripetere vertici intermedi. Ma ogni foglia ha un solo arco (verso $C$): un cammino che entra in una foglia non può proseguire altrove se non tornando indietro per lo stesso arco (il che non è ammesso in un ciclo, che non ripete archi... in realtà non ripete vertici). Non essendoci archi tra le foglie, l'unico modo di "muoversi" è passare sempre per $C$, e per formare un ciclo servirebbe rientrare in $C$ due volte, il che non è permesso (un ciclo visita ogni vertice al più una volta, escluso l'estremo). Nessun ciclo possibile: è aciclico.

**Passo 3 — conclusione.** Connesso e aciclico: è un albero, per definizione. Ha $n=5$ vertici ($C,F_1,F_2,F_3,F_4$) e $4$ archi ($C$-$F_1$, $C$-$F_2$, $C$-$F_3$, $C$-$F_4$): infatti $4=n-1=5-1$, coerente col teorema.


```python
import numpy as np

vertici = ["C", "F1", "F2", "F3", "F4"]
n = len(vertici)
archi = [("C", "F1"), ("C", "F2"), ("C", "F3"), ("C", "F4")]

idx = {v: i for i, v in enumerate(vertici)}
A = np.zeros((n, n), dtype=int)
for u, v in archi:
    A[idx[u], idx[v]] = 1
    A[idx[v], idx[u]] = 1

print(f"n = {n} vertici, |E| = {len(archi)} archi, n - 1 = {n - 1}")
print(f"|E| == n - 1 ? {len(archi) == n - 1}")

gradi = A.sum(axis=1)
print(f"Gradi: {dict(zip(vertici, gradi))}")
print(f"Somma gradi = {gradi.sum()} = 2*|E| = {2 * len(archi)}  (lemma delle strette di mano, verificato anche qui)")
```

## Riepilogo finale

- Un **grafo** $G=(V,E)$ modella oggetti (vertici) e collegamenti (archi); il **grado** di un vertice conta gli archi che lo toccano.
- Il **lemma delle strette di mano** ($\sum_v\deg(v)=2|E|$) si dimostra per doppio conteggio: la stessa tecnica usata più volte in questo modulo.
- Un **albero** (connesso, aciclico) con $n$ vertici ha esattamente $n-1$ archi — dimostrato per induzione rimuovendo una foglia alla volta.
- La **matrice di adiacenza** $A$ codifica il grafo algebricamente: $(A^k)_{ij}$ conta i cammini di lunghezza $k$ da $i$ a $j$, un collegamento diretto con le potenze di matrici del Modulo 6.
- Il problema dei **ponti di Königsberg** (Eulero, 1736), risolto mostrando che un circuito che percorre ogni arco una volta richiede gradi tutti pari, è il primo teorema della teoria dei grafi e un modello di dimostrazione di non esistenza.
