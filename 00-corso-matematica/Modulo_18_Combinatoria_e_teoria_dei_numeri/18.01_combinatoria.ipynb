## Perché ci interessa

Nell'ultimo esercizio del notebook 17.10 abbiamo incontrato una domanda molto concreta: in quanti modi diversi una passeggiata aleatoria di $4$ passi può arrivare in posizione $S_4=2$? Abbiamo risposto elencando a mano le quattro sequenze possibili, e abbiamo notato che il risultato coincide con un coefficiente binomiale, $\binom43=4$. Ma elencare a mano funziona solo per numeri piccoli: per $100$ passi sarebbe impossibile. Serve un modo **sistematico** di contare, senza elencare.

Questo è l'oggetto della **combinatoria**: contare quante sono le configurazioni possibili di un certo tipo, in modo esatto, usando principi generali invece dell'elenco esplicito. Non è un argomento isolato: il conteggio è alla base della probabilità su spazi finiti (Modulo 17, dove ogni probabilità $P(A)=|A|/|\Omega|$ nel caso equiprobabile è, letteralmente, un rapporto di conteggi), e riapparirà in ogni angolo della matematica discreta che vedremo in questo modulo conclusivo del corso.


## Teoria, passo dopo passo

### Il principio fondamentale del conteggio

Immaginiamo di dover compiere una scelta in **due fasi successive**: la prima fase ha $m$ esiti possibili, e per ciascuno di essi la seconda fase ne ha $n$, indipendentemente da quale esito abbia avuto la prima. Quante sono le coppie (esito fase 1, esito fase 2) possibili?

Per ogni scelta della prima fase, esistono esattamente $n$ modi di completare la seconda: quindi il totale è $m$ gruppi da $n$ ciascuno, cioè $m\cdot n$. Questo è il **principio moltiplicativo**, e si estende a $k$ fasi successive: se la fase $i$ ha $n_i$ esiti (indipendentemente dalle fasi precedenti), il numero totale di sequenze di scelte è $n_1\cdot n_2\cdots n_k$. Lo useremo continuamente, spesso senza nemmeno nominarlo esplicitamente.

### Permutazioni: ordinare $n$ oggetti distinti

**Problema guida:** in quanti modi possiamo disporre in fila $n$ oggetti tutti distinti?

Pensiamo di riempire $n$ posizioni una alla volta. Per la prima posizione abbiamo $n$ oggetti disponibili. Una volta scelto quello, per la seconda posizione ne restano $n-1$ (uno è già stato usato). Per la terza, $n-2$. E così via, fino all'ultima posizione, per cui resta un solo oggetto. Per il principio moltiplicativo, il numero di disposizioni in fila (le **permutazioni** di $n$ oggetti) è
$$n\cdot(n-1)\cdot(n-2)\cdots2\cdot1=:n!$$
(si legge "$n$ fattoriale"). Per convenzione $0!=1$ (c'è esattamente un modo di disporre in fila zero oggetti: non fare nulla).

### Disposizioni: scegliere e ordinare $k$ oggetti su $n$

**Problema guida:** su $n$ atleti distinti, in quanti modi possiamo assegnare oro, argento e bronzo (3 premi distinti, nessun atleta ne riceve più di uno)?

Stesso ragionamento di prima, ma ci fermiamo dopo $k=3$ posizioni invece di $n$: $n$ scelte per l'oro, $n-1$ per l'argento, $n-2$ per il bronzo. In generale, il numero di modi di scegliere **e ordinare** $k$ oggetti distinti tra $n$ disponibili (le **disposizioni semplici** di $n$ oggetti in classe $k$, con $0\le k\le n$) è
$$D(n,k)=n\cdot(n-1)\cdots(n-k+1)=\frac{n!}{(n-k)!}.$$
Verifica: per $k=n$ si ha $D(n,n)=n!/0!=n!$, coerente con le permutazioni.

**Disposizioni con ripetizione.** Se invece uno stesso oggetto può essere scelto più volte (es. una password di $k$ caratteri presi da un alfabeto di $n$ simboli, ripetizioni ammesse), ogni posizione ha sempre $n$ scelte indipendenti dalle altre: per il principio moltiplicativo il totale è semplicemente $n^k$.

### Combinazioni: scegliere $k$ oggetti su $n$, senza ordinarli

**Problema guida:** su $n$ persone, in quanti modi possiamo formare un comitato di $k$ persone (nessun ruolo distinto, solo "fa parte del comitato" o no)?

Partiamo da $D(n,k)$, che conta i comitati **ordinati** (come se assegnassimo un ordine di elezione). Ogni comitato non ordinato di $k$ persone corrisponde esattamente a $k!$ comitati ordinati (tutti i modi di ordinare le stesse $k$ persone). Quindi il numero di comitati non ordinati — le **combinazioni** di $n$ oggetti in classe $k$ — è
$$\binom nk:=\frac{D(n,k)}{k!}=\frac{n!}{k!(n-k)!}.$$
Questo è il **coefficiente binomiale**, già incontrato nel notebook 17.10. Vale $\binom nk=\binom n{n-k}$ (scegliere $k$ elementi da includere equivale a scegliere $n-k$ elementi da escludere: una biiezione tra le due famiglie di sottoinsiemi, nel senso del notebook 3.5).

**Relazione di Pascal.** Fissiamo un elemento particolare, diciamo l'ultimo, tra gli $n$. Un sottoinsieme di $k$ elementi o contiene quell'elemento oppure no:
- se lo contiene, restano da scegliere $k-1$ elementi tra i rimanenti $n-1$: $\binom{n-1}{k-1}$ modi;
- se non lo contiene, restano da scegliere tutti e $k$ gli elementi tra i rimanenti $n-1$: $\binom{n-1}k$ modi.

Questi due casi sono disgiunti ed esauriscono tutte le possibilità, quindi
$$\binom nk=\binom{n-1}{k-1}+\binom{n-1}k,$$
la regola che genera il **triangolo di Tartaglia (o di Pascal)**: ogni voce è la somma delle due sopra di essa.

### Il binomio di Newton

**Teorema.** Per ogni $n\ge0$ e ogni $x,y$,
$$(x+y)^n=\sum_{k=0}^n\binom nkx^ky^{n-k}.$$

**Dimostrazione (per induzione su $n$, notebook 10.4).**

*Base $n=0$:* $(x+y)^0=1$, e il membro destro è $\binom00x^0y^0=1$. Vero.

*Passo induttivo.* Supponiamo la formula vera per $n$ (ipotesi induttiva), e mostriamola per $n+1$:
$$(x+y)^{n+1}=(x+y)(x+y)^n\overset{\text{ip. ind.}}{=}(x+y)\sum_{k=0}^n\binom nkx^ky^{n-k}=\sum_{k=0}^n\binom nkx^{k+1}y^{n-k}+\sum_{k=0}^n\binom nkx^ky^{n-k+1}.$$
Nella prima somma poniamo $j=k+1$ (quindi $k=j-1$, e $j$ va da $1$ a $n+1$); nella seconda lasciamo $j=k$ (da $0$ a $n$):
$$=\sum_{j=1}^{n+1}\binom n{j-1}x^jy^{n+1-j}+\sum_{j=0}^n\binom njx^jy^{n+1-j}.$$
Isoliamo il termine $j=n+1$ dalla prima somma e $j=0$ dalla seconda, e sommiamo i termini comuni $j=1,\ldots,n$ usando la relazione di Pascal $\binom n{j-1}+\binom nj=\binom{n+1}j$:
$$=x^{n+1}+\sum_{j=1}^n\Big(\binom n{j-1}+\binom nj\Big)x^jy^{n+1-j}+y^{n+1}=x^{n+1}+\sum_{j=1}^n\binom{n+1}jx^jy^{n+1-j}+y^{n+1}=\sum_{j=0}^{n+1}\binom{n+1}jx^jy^{n+1-j}.$$
(l'ultimo passaggio inglobando $x^{n+1}=\binom{n+1}{n+1}x^{n+1}y^0$ e $y^{n+1}=\binom{n+1}0x^0y^{n+1}$ nella somma). Questo è esattamente la formula per $n+1$. $\blacksquare$

**Perché ha senso il nome "coefficiente binomiale".** Il teorema spiega perché $\binom nk$ compare come coefficiente nello sviluppo di una potenza di un binomio — non è una coincidenza di notazione, ma il motivo storico del nome.

### Permutazioni con ripetizione (anagrammi)

**Problema guida:** quanti anagrammi (distinti come stringhe) ha la parola "MATEMATICA" (10 lettere, di cui M compare 2 volte, A 3 volte, T 2 volte, E 1 volta, I 1 volta, C 1 volta)?

Se le $10$ lettere fossero tutte distinte, avremmo $10!$ permutazioni. Ma scambiare tra loro le $2$ M non produce una stringa diversa: ogni anagramma "vero" viene contato $2!$ volte in più per le M, $3!$ volte in più per le A, $2!$ volte in più per le T. Correggendo per questa sovraconta (lo stesso principio userà in modo più sistematico il notebook 18.2):
$$\frac{10!}{2!\,3!\,2!\,1!\,1!\,1!}.$$
In generale, il numero di permutazioni distinte di $n$ oggetti di cui $n_1$ uguali tra loro, $n_2$ altri uguali tra loro, ..., $n_r$ uguali tra loro (con $n_1+\cdots+n_r=n$) è il **coefficiente multinomiale**
$$\binom n{n_1,n_2,\ldots,n_r}=\frac{n!}{n_1!\,n_2!\cdots n_r!}.$$


```python
from itertools import permutations, combinations, product
from math import factorial, comb

# Verifichiamo le formule D(n,k) e C(n,k) = binom(n,k) per via enumerativa
# (elenco esplicito), confrontando con le formule chiuse.

n, k = 6, 3

disposizioni = list(permutations(range(n), k))
combinazioni = list(combinations(range(n), k))

print(f"n={n}, k={k}")
print(f"D(n,k) enumerato = {len(disposizioni)}, formula n!/(n-k)! = {factorial(n)//factorial(n-k)}")
print(f"C(n,k) enumerato = {len(combinazioni)}, formula comb(n,k) = {comb(n, k)}")

# Ogni combinazione corrisponde a k! disposizioni: verifichiamolo direttamente
rapporto = len(disposizioni) / len(combinazioni)
print(f"D(n,k)/C(n,k) = {rapporto}, k! = {factorial(k)}")

# Disposizioni con ripetizione: contiamo con product() e confrontiamo con n**k
disposizioni_rip = list(product(range(n), repeat=k))
print(f"\nDisposizioni con ripetizione enumerate = {len(disposizioni_rip)}, formula n^k = {n**k}")

# Triangolo di Pascal: costruito riga per riga con la relazione binom(n,k)=binom(n-1,k-1)+binom(n-1,k)
righe = 8
triangolo = [[comb(r, c) for c in range(r + 1)] for r in range(righe)]
for r, riga in enumerate(triangolo):
    print(" " * (righe - r), *riga)
```

## Esercizio svolto e commentato

**Problema:** una squadra di calcetto ha $12$ giocatori disponibili. Il modulo di gioco richiede $5$ giocatori in campo, tutti intercambiabili tranne il portiere (un ruolo specifico, che deve essere ricoperto da uno dei $3$ giocatori che sanno parare). Quanti gruppi di $5$ giocatori (portiere + 4 di movimento) si possono formare?

**Ragionamento:**

**Passo 1 — separiamo le due scelte indipendenti.** La scelta del portiere e la scelta dei $4$ giocatori di movimento sono due fasi indipendenti nel senso del principio moltiplicativo: il portiere si sceglie tra i $3$ specialisti, i $4$ di movimento tra i restanti $9$ giocatori (i $12$ totali meno i $3$ portieri, perché un giocatore di movimento non può essere anche il portiere in questa formazione).

**Passo 2 — contiamo la scelta del portiere.** È una scelta singola tra $3$ candidati, senza ulteriore struttura: $3$ modi.

**Passo 3 — contiamo la scelta dei $4$ giocatori di movimento.** Non c'è un ordine tra i $4$ (sono "intercambiabili" per ipotesi): è una combinazione, non una disposizione. Quindi $\binom94$ modi.

**Passo 4 — applichiamo il principio moltiplicativo.** Il numero totale di gruppi è
$$3\cdot\binom94=3\cdot126=378.$$

**Perché non abbiamo usato una disposizione per i 4 di movimento?** Perché il problema dice esplicitamente che sono intercambiabili: la squadra $\{a,b,c,d\}$ in campo è la stessa formazione qualunque sia l'ordine in cui elenchiamo i quattro nomi. Se invece i $4$ ruoli di movimento fossero specifici e distinti (es. due difensori, un centrocampista, un attaccante, ciascuno un ruolo diverso), servirebbe una disposizione o un conteggio a fasi separate — un errore comune è usare $D(n,k)$ quando l'ordine non conta, o viceversa: la domanda da farsi sempre è **"due scelte con gli stessi elementi ma in ordine diverso sono la stessa configurazione, o no?"**


```python
from math import comb

n_portieri = 3
n_movimento_disponibili = 12 - 3   # i 9 giocatori che non sono specialisti di porta
k_movimento = 4

modi_portiere = n_portieri
modi_movimento = comb(n_movimento_disponibili, k_movimento)
totale = modi_portiere * modi_movimento

print(f"Modi di scegliere il portiere: {modi_portiere}")
print(f"Modi di scegliere i 4 di movimento: C(9,4) = {modi_movimento}")
print(f"Totale gruppi possibili: {totale}")
```

## Riepilogo finale

- Il **principio moltiplicativo** (fasi indipendenti si moltiplicano) e quello **additivo** (casi disgiunti si sommano) sono i due mattoni di ogni conteggio.
- **Permutazioni** ($n!$): ordinare $n$ oggetti distinti. **Disposizioni** $D(n,k)=n!/(n-k)!$: scegliere e ordinare $k$ oggetti su $n$. **Combinazioni** $\binom nk=n!/(k!(n-k)!)$: scegliere $k$ oggetti su $n$ **senza** ordinarli — la domanda chiave è sempre "l'ordine conta?".
- Il **binomio di Newton** $(x+y)^n=\sum_k\binom nkx^ky^{n-k}$ spiega perché $\binom nk$ si chiama "coefficiente binomiale", e si dimostra per induzione usando la relazione di Pascal $\binom nk=\binom{n-1}{k-1}+\binom{n-1}k$.
- I **coefficienti multinomiali** generalizzano le combinazioni al conteggio di permutazioni con ripetizioni (anagrammi con lettere ripetute).
- Il conteggio esatto funziona bene finché gli insiemi che vogliamo contare sono "puliti" (nessuna sovrapposizione tra casi); quando i casi si sovrappongono, serve correggere la sovraconta — il problema del prossimo notebook.
