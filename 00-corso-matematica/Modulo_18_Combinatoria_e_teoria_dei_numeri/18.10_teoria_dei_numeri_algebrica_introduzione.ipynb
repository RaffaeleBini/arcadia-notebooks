## Perché ci interessa

Nel notebook 18.7 abbiamo incontrato senza dimostrarla la congettura di Goldbach; qui affrontiamo invece un problema altrettanto naturale ma **risolubile** con gli strumenti giusti: **quali numeri primi si scrivono come somma di due quadrati**? Per esempio $5=1^2+2^2$ e $13=2^2+3^2$ funzionano, ma $7$ e $11$ no, per quanti tentativi si facciano. C'è un pattern?

La teoria dei numeri elementare (Moduli 18.5-18.9), lavorando solo dentro $\mathbb Z$, non ha gli strumenti per rispondere a questa domanda in modo diretto. Ma se ampliamo l'ambiente numerico — non fino a $\mathbb R$ o $\mathbb C$ per intero, ma a un anello intermedio, gli **interi di Gauss** $\mathbb Z[i]$ — la domanda diventa una questione di fattorizzazione, esattamente come nel Modulo 18.5, e la risposta emerge in modo naturale. Questo notebook, capstone conclusivo dell'intero corso, mostra come algebra astratta (Modulo 11), numeri complessi (Modulo 9) e teoria dei numeri (questo modulo) si fondano in un'unica idea: la **teoria dei numeri algebrica**, il punto d'incontro di tutte e tre.


## Teoria, passo dopo passo

### Gli interi di Gauss

Gli **interi di Gauss** sono l'insieme
$$\mathbb Z[i]=\{a+bi:a,b\in\mathbb Z\}\subset\mathbb C$$
(notebook 9.1: $i^2=-1$). Con le usuali operazioni di somma e prodotto dei numeri complessi, $\mathbb Z[i]$ è un **anello commutativo** (notebook 11.6): la somma e il prodotto di due interi di Gauss restano interi di Gauss (verifica diretta: $(a+bi)+(c+di)=(a+c)+(b+d)i$, $(a+bi)(c+di)=(ac-bd)+(ad+bc)i$, e tutte le componenti restano intere), con gli stessi assiomi (associatività, distributività, ecc.) ereditati da $\mathbb C$.

### La norma

La **norma** di $z=a+bi\in\mathbb Z[i]$ è $N(z)=a^2+b^2=z\bar z=|z|^2$ (il quadrato del modulo, notebook 9.1-9.2) — sempre un intero non negativo.

**Proposizione (moltiplicatività della norma).** $N(zw)=N(z)N(w)$ per ogni $z,w\in\mathbb Z[i]$.

**Dimostrazione.** $N(zw)=(zw)\overline{(zw)}=zw\bar z\bar w=(z\bar z)(w\bar w)=N(z)N(w)$, usando $\overline{zw}=\bar z\bar w$ (proprietà del coniugato, notebook 9.1) e la commutatività del prodotto. $\blacksquare$

Questa identità, scritta per esteso con $z=a+bi,w=c+di$, dà l'**identità di Brahmagupta-Fibonacci**:
$$(a^2+b^2)(c^2+d^2)=(ac-bd)^2+(ad+bc)^2,$$
nota secoli prima della sua interpretazione via numeri complessi: il prodotto di due somme di due quadrati è ancora una somma di due quadrati — un fatto che, ora, è solo la moltiplicatività della norma travestita.

### Le unità di $\mathbb Z[i]$

Un elemento $u\in\mathbb Z[i]$ è un'**unità** se ha un inverso moltiplicativo in $\mathbb Z[i]$ (notebook 11.10, ma qui $\mathbb Z[i]$ non è un campo, quindi non tutti gli elementi non nulli sono invertibili — solo alcuni, come in $\mathbb Z$ dove le uniche unità sono $\pm1$).

**Proposizione.** Le unità di $\mathbb Z[i]$ sono esattamente $\{1,-1,i,-i\}$.

**Dimostrazione.** Se $u$ è un'unità, esiste $v$ con $uv=1$; per moltiplicatività della norma, $N(u)N(v)=N(1)=1$; essendo $N(u),N(v)$ interi non negativi con prodotto $1$, deve essere $N(u)=N(v)=1$. Scrivendo $u=a+bi$, $N(u)=a^2+b^2=1$ con $a,b\in\mathbb Z$: le uniche soluzioni sono $(a,b)=(\pm1,0)$ o $(0,\pm1)$, cioè $u\in\{1,-1,i,-i\}$. Viceversa, ciascuno di questi quattro elementi è effettivamente invertibile ($1\cdot1=1$, $(-1)(-1)=1$, $i\cdot(-i)=1$): sono tutte e sole le unità. $\blacksquare$

### Divisione con resto in $\mathbb Z[i]$

**Teorema (senza dimostrazione completa).** Per ogni $z,w\in\mathbb Z[i]$ con $w\ne0$, esistono $q,r\in\mathbb Z[i]$ con $z=qw+r$ e $N(r)<N(w)$.

*(Idea: si calcola $z/w\in\mathbb C$ esattamente come numero complesso, poi si arrotonda ciascuna componente all'intero più vicino per ottenere $q$; il resto $r=z-qw$ ha norma strettamente minore di $N(w)$ perché l'arrotondamento introduce un errore di al più $1/2$ in ciascuna componente, e $(1/2)^2+(1/2)^2=1/2<1$.)*

Questo teorema rende $\mathbb Z[i]$ un **dominio euclideo** (una generalizzazione, non approfondita nel Modulo 11, della struttura di $\mathbb Z$ vista nel notebook 18.5): esattamente come in $\mathbb Z$, si può quindi costruire un **algoritmo di Euclide** in $\mathbb Z[i]$, dimostrare un'identità di Bézout, e — conseguenza finale — un **teorema fondamentale dell'aritmetica** per $\mathbb Z[i]$: ogni intero di Gauss non nullo e non unità si fattorizza in modo unico (a meno di riordino e di moltiplicazione per unità) in **primi di Gauss**.

### Primi di Gauss e primi ordinari

Un primo ordinario $p\in\mathbb Z$ può comportarsi in due modi diversi in $\mathbb Z[i]$:

- **resta primo** in $\mathbb Z[i]$ (nessuna fattorizzazione propria), oppure
- **si fattorizza** come prodotto di due primi di Gauss coniugati, $p=\pi\bar\pi$, con $N(\pi)=p$.

**Proposizione.** Se $p\equiv3\pmod4$, allora $p$ resta primo in $\mathbb Z[i]$.

**Dimostrazione.** Supponiamo per assurdo $p=zw$ con $z,w\in\mathbb Z[i]$ non unità. Allora $N(p)=p^2=N(z)N(w)$, con $N(z),N(w)>1$ (nessuno dei due è un'unità): quindi $N(z)=N(w)=p$ (essendo $p$ primo in $\mathbb Z$, l'unica fattorizzazione propria di $p^2$ in due fattori $>1$ è $p\cdot p$). Scrivendo $z=a+bi$, questo significa $a^2+b^2=p$. Ma i quadrati modulo $4$ sono solo $0$ o $1$ (essendo $(2k)^2=4k^2\equiv0$ e $(2k+1)^2=4k^2+4k+1\equiv1\pmod4$), quindi $a^2+b^2\bmod4\in\{0,1,2\}$ — mai $3$. Questo contraddice $p\equiv3\pmod4$. Quindi nessuna fattorizzazione propria esiste: $p$ resta primo. $\blacksquare$

*(Il caso $p\equiv1\pmod4$, in cui $p$ **si fattorizza** in $\mathbb Z[i]$, richiede uno strumento — l'esistenza di una radice quadrata di $-1$ modulo $p$ — che enunciamo senza dimostrare: per $p\equiv1\pmod4$, esiste $m$ con $m^2\equiv-1\pmod p$; allora $p\mid(m^2+1)=(m+i)(m-i)$ in $\mathbb Z[i]$, ma $p$ non divide né $m+i$ né $m-i$ singolarmente in $\mathbb Z[i]$ (nessuno dei due quozienti $\frac{m\pm i}p$ ha componenti intere), quindi $p$ non può essere primo in $\mathbb Z[i]$ per il lemma di Euclide generalizzato — e si fattorizza.)*

### Il teorema dei due quadrati di Fermat

**Teorema.** Un primo dispari $p$ si scrive come somma di due quadrati, $p=a^2+b^2$, se e solo se $p\equiv1\pmod4$.

**Dimostrazione della direzione facile ($\Rightarrow$).** Se $p=a^2+b^2$, allora modulo $4$: i quadrati sono $0$ o $1$ (come sopra), quindi $p\bmod4\in\{0,1,2\}$. Essendo $p$ primo dispari, $p\not\equiv0\pmod4$ (pari) e $p\not\equiv2\pmod4$ (pari); resta $p\equiv1\pmod4$.

**Direzione difficile ($\Leftarrow$, solo idea, collegata alla teoria sopra).** Se $p\equiv1\pmod4$, per il fatto enunciato sopra $p$ si fattorizza in $\mathbb Z[i]$ come $p=\pi\bar\pi$ con $\pi=a+bi$, $N(\pi)=a^2+b^2=p$: la fattorizzazione di $p$ in $\mathbb Z[i]$ **è**, letteralmente, la rappresentazione di $p$ come somma di due quadrati. $\blacksquare$ (nella parte che dimostriamo per esteso)

Questo è il senso in cui la teoria dei numeri algebrica "risolve" un problema che la teoria elementare non può affrontare da sola: la domanda su $\mathbb Z$ ("chi è somma di due quadrati?") diventa una domanda di fattorizzazione in $\mathbb Z[i]$ — un problema dello stesso tipo del teorema fondamentale dell'aritmetica (notebook 18.5), solo in un anello più ricco.

### Uno sguardo alla teoria dei numeri algebrica, e la chiusura del corso

$\mathbb Z[i]$ è il primo esempio di **anello di interi algebrici**: l'idea si generalizza ad altri anelli $\mathbb Z[\omega]$ (con $\omega$ radice di un polinomio a coefficienti interi, come le radici dell'unità del notebook 9.3), ciascuno con la propria aritmetica, non sempre a fattorizzazione unica (un fenomeno scoperto proprio studiando queste generalizzazioni, e storicamente uno dei motori che ha portato alla nascita della moderna teoria degli anelli, Modulo 11) — un intero campo di ricerca attivo ancora oggi, ben oltre lo scopo di questo corso introduttivo.

Con questo notebook si chiude l'intero percorso iniziato al notebook 0.1 con i numeri naturali. Il cerchio si chiude in modo naturale: siamo partiti chiedendoci cosa fosse un numero, e finiamo chiedendoci come i numeri si fattorizzano in strutture via via più ricche — $\mathbb N\to\mathbb Z\to\mathbb Q\to\mathbb R\to\mathbb C$ (Modulo 0, 9) — usando nel percorso l'algebra (Moduli 1, 6, 11), l'analisi (Moduli 4, 5, 7, 13, 14, 15), la geometria e la topologia (Moduli 2, 12, 16), la probabilità (Modulo 17), e infine la combinatoria e la teoria dei numeri di questo ultimo modulo — tutte parti di un unico edificio.


```python
def norma(a, b):
    return a * a + b * b

def fattorizza_gauss_p_equiv_1_mod_4(p):
    # Per p primo con p = 1 mod 4, trova a,b interi con a^2+b^2=p per tentativi
    # (metodo elementare, non efficiente ma sufficiente per verificare la teoria).
    for a in range(1, int(p**0.5) + 1):
        b2 = p - a * a
        b = round(b2**0.5)
        if b >= 0 and b * b == b2:
            return a, b
    return None

primi_da_testare = [5, 13, 17, 29, 37, 41]
print("Primi p = 1 mod 4: rappresentazione come somma di due quadrati")
for p in primi_da_testare:
    assert p % 4 == 1
    ab = fattorizza_gauss_p_equiv_1_mod_4(p)
    a, b = ab
    print(f"  {p} = {a}^2 + {b}^2 = {a*a} + {b*b}  (verifica: {norma(a,b)} == {p} ? {norma(a,b) == p})")

primi_3_mod_4 = [3, 7, 11, 19, 23]
print("\nPrimi p = 3 mod 4: nessuna rappresentazione come somma di due quadrati")
for p in primi_3_mod_4:
    assert p % 4 == 3
    trovato = any(a*a + b*b == p for a in range(int(p**0.5) + 1) for b in range(int(p**0.5) + 1))
    print(f"  {p}: rappresentazione trovata? {trovato}  (atteso: False)")

# Verifica dell'identita' di Brahmagupta-Fibonacci: (a^2+b^2)(c^2+d^2) = (ac-bd)^2+(ad+bc)^2
a, b, c, d = 1, 2, 2, 3
sinistra = (a*a + b*b) * (c*c + d*d)
destra = (a*c - b*d)**2 + (a*d + b*c)**2
print(f"\nIdentita' di Brahmagupta-Fibonacci: ({a}^2+{b}^2)({c}^2+{d}^2) = {sinistra},  (ac-bd)^2+(ad+bc)^2 = {destra}")
```

## Esercizio svolto e commentato

**Problema:** fattorizza $13$ come prodotto di due primi di Gauss coniugati (sapendo $13\equiv1\pmod4$), e verifica la moltiplicatività della norma sulla fattorizzazione trovata.

**Ragionamento:**

**Passo 1 — verifichiamo $13\equiv1\pmod4$.** $13=3\cdot4+1$: sì, quindi per la teoria $13$ si fattorizza in $\mathbb Z[i]$.

**Passo 2 — cerchiamo $a,b$ con $a^2+b^2=13$.** Per tentativi: $a=1\Rightarrow b^2=12$ (non quadrato); $a=2\Rightarrow b^2=9=3^2$ ✓. Quindi $a=2,b=3$: $2^2+3^2=4+9=13$.

**Passo 3 — scriviamo la fattorizzazione.** $\pi=2+3i$, $\bar\pi=2-3i$, e $\pi\bar\pi=N(\pi)=13$: quindi $13=(2+3i)(2-3i)$.

**Passo 4 — verifichiamo esplicitamente il prodotto.** $(2+3i)(2-3i)=2^2-(3i)^2=4-9i^2=4-9(-1)=4+9=13$ ✓ (usando la differenza di quadrati, notebook 1.2, applicata ai numeri complessi).

**Conclusione (verificare nella cella seguente):** $13$, primo in $\mathbb Z$, **non** è primo in $\mathbb Z[i]$: si fattorizza nei due primi di Gauss coniugati $2+3i$ e $2-3i$, entrambi di norma $13$.


```python
pi = complex(2, 3)
pi_coniugato = complex(2, -3)
prodotto = pi * pi_coniugato

print(f"pi = {pi}, pi_coniugato = {pi_coniugato}")
print(f"pi * pi_coniugato = {prodotto}")
print(f"Parte reale (dovrebbe essere 13): {prodotto.real}")
print(f"Parte immaginaria (dovrebbe essere 0): {prodotto.imag}")

# Verifica con sympy usando I (unita' immaginaria simbolica, aritmetica esatta)
from sympy import I, expand
pi_sym = 2 + 3*I
pi_coniugato_sym = 2 - 3*I
prodotto_sym = expand(pi_sym * pi_coniugato_sym)
print(f"\nCon aritmetica esatta (sympy): (2+3i)(2-3i) = {prodotto_sym}")
```

## Riepilogo finale

- Gli **interi di Gauss** $\mathbb Z[i]=\{a+bi:a,b\in\mathbb Z\}$ formano un anello commutativo con **norma moltiplicativa** $N(a+bi)=a^2+b^2$, le cui uniche unità sono $\pm1,\pm i$.
- $\mathbb Z[i]$ ammette divisione con resto (dominio euclideo), quindi — come $\mathbb Z$ — un algoritmo di Euclide e un teorema fondamentale dell'aritmetica con **primi di Gauss**.
- Un primo ordinario $p\equiv3\pmod4$ resta primo in $\mathbb Z[i]$; un primo $p\equiv1\pmod4$ si fattorizza come $\pi\bar\pi$.
- Il **teorema dei due quadrati di Fermat** ($p=a^2+b^2\iff p\equiv1\pmod4$, per $p$ primo dispari) traduce un problema elementare su $\mathbb Z$ in un problema di fattorizzazione su $\mathbb Z[i]$: l'idea guida di tutta la teoria dei numeri algebrica.
- $\mathbb Z[i]$ è il primo passo verso gli **anelli di interi algebrici** più generali, un campo di ricerca tuttora attivo.
