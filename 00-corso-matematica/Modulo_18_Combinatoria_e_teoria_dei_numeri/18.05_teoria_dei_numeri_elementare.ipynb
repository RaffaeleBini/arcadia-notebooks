## Teoria, passo dopo passo

### Divisibilità

Dati $a,b\in\mathbb Z$, diciamo che $a$ **divide** $b$ (scritto $a\mid b$) se esiste $q\in\mathbb Z$ tale che $b=aq$. Proprietà immediate (dimostrazioni dirette, lasciate come esercizio mentale): $a\mid a$ (riflessività, con $q=1$); se $a\mid b$ e $b\mid c$ allora $a\mid c$ (transitività: $b=aq_1,c=bq_2\Rightarrow c=aq_1q_2$); se $a\mid b$ e $a\mid c$ allora $a\mid(bx+cy)$ per ogni $x,y\in\mathbb Z$ (perché $bx+cy=a(q_1x+q_2y)$) — quest'ultima proprietà, la **chiusura per combinazioni lineari intere**, sarà cruciale tra poco.

### Il teorema della divisione euclidea

**Teorema.** Per ogni $a\in\mathbb Z$ e $b\in\mathbb Z$ con $b>0$, esistono unici $q,r\in\mathbb Z$ tali che
$$a=bq+r,\qquad0\le r<b.$$

**Dimostrazione dell'esistenza.** Consideriamo l'insieme $S=\{a-bq:q\in\mathbb Z,\ a-bq\ge0\}$. $S$ è non vuoto (prendendo $q$ molto negativo se $a\ge0$ serve poco, ma in generale: se $a\ge0$, $q=0$ dà $a-b\cdot0=a\ge0\in S$; se $a<0$, $q=a$ dà $a-ba=a(1-b)\ge0$ perché $b\ge1\Rightarrow1-b\le0$ e $a<0$, quindi il prodotto di due numeri $\le0$ è $\ge0$) e contenuto in $\mathbb Z_{\ge0}$. Per il **principio del buon ordinamento** (ogni sottoinsieme non vuoto di $\mathbb Z_{\ge0}$ ha un minimo, la versione intera della completezza vista nel notebook 5.2), $S$ ha un elemento minimo $r=a-bq$. Mostriamo $r<b$: se fosse $r\ge b$, allora $r-b=a-b(q+1)\ge0$ apparterrebbe a $S$ ed sarebbe strettamente minore di $r$ (essendo $b>0$), contraddicendo la minimalità di $r$. Quindi $0\le r<b$.

**Dimostrazione dell'unicità.** Supponiamo $a=bq+r=bq'+r'$ con $0\le r,r'<b$. Allora $b(q-q')=r'-r$, quindi $b\mid(r'-r)$. Ma $|r'-r|<b$ (essendo entrambi in $[0,b)$), e l'unico multiplo di $b$ con valore assoluto $<b$ è $0$: quindi $r'=r$, e di conseguenza $b(q-q')=0$ con $b\ne0$, cioè $q=q'$. $\blacksquare$

Questo teorema formalizza la divisione con resto insegnata a scuola (notebook 0.2), e la sua dimostrazione — trovare il minimo di un insieme ben ordinato — è lo schema che ritroveremo tra poco per l'algoritmo di Euclide.

### Massimo comun divisore e algoritmo di Euclide

Il **massimo comun divisore** $\gcd(a,b)$ (con $a,b$ non entrambi nulli) è il più grande intero positivo che divide sia $a$ che $b$.

**Lemma.** Se $a=bq+r$ (divisione euclidea), allora $\gcd(a,b)=\gcd(b,r)$.

**Dimostrazione.** Mostriamo che $a,b$ e $b,r$ hanno esattamente gli stessi divisori comuni, quindi in particolare lo stesso massimo. Se $d\mid a$ e $d\mid b$, allora $d\mid(a-bq)=r$ (chiusura per combinazioni lineari, vista sopra): quindi $d$ divide anche $b$ e $r$. Viceversa, se $d\mid b$ e $d\mid r$, allora $d\mid(bq+r)=a$: quindi $d$ divide anche $a$ e $b$. I due insiemi di divisori comuni coincidono, quindi i due massimi coincidono. $\blacksquare$

**L'algoritmo di Euclide** applica questo lemma ripetutamente: $\gcd(a,b)=\gcd(b,r_1)=\gcd(r_1,r_2)=\cdots$, dove $r_1$ è il resto di $a$ diviso $b$, $r_2$ il resto di $b$ diviso $r_1$, e così via. I resti $r_1>r_2>\cdots\ge0$ formano una successione strettamente decrescente di interi non negativi, quindi (di nuovo per il buon ordinamento) deve raggiungere $0$ in un numero finito di passi: l'ultimo resto non nullo è $\gcd(a,b)$.

### L'identità di Bézout

**Teorema.** Per ogni $a,b\in\mathbb Z$ non entrambi nulli, esistono $x,y\in\mathbb Z$ tali che $ax+by=\gcd(a,b)$.

**Dimostrazione.** Consideriamo l'insieme $S=\{ax+by:x,y\in\mathbb Z\}\cap\mathbb Z_{>0}$ (le combinazioni lineari intere positive di $a,b$). $S$ è non vuoto (es. $a\cdot a+b\cdot b=a^2+b^2>0$, essendo $a,b$ non entrambi nulli), quindi per il buon ordinamento ha un minimo $d=ax_0+by_0$. Mostriamo $d=\gcd(a,b)$.

*$d$ divide $a$:* dividiamo $a=dq+r$ con $0\le r<d$ (divisione euclidea). Allora $r=a-dq=a-(ax_0+by_0)q=a(1-x_0q)+b(-y_0q)$, cioè $r$ è anch'esso una combinazione lineare intera di $a,b$. Se fosse $r>0$, apparterrebbe a $S$ ed sarebbe minore di $d$, contraddicendo la minimalità: quindi $r=0$, cioè $d\mid a$. Simmetricamente, $d\mid b$.

*$d$ è il **massimo** tra i divisori comuni:* se $c$ è un qualunque divisore comune di $a,b$, allora $c\mid(ax_0+by_0)=d$ (chiusura per combinazioni lineari), quindi $c\le d$ (essendo $d>0$). Quindi $d$ è divisibile da ogni divisore comune ed è esso stesso un divisore comune: è il massimo. $\blacksquare$

**Osservazione.** L'algoritmo di Euclide "esteso" (che teniamo traccia dei coefficienti a ogni passo, non solo dei resti) calcola esplicitamente $x,y$: lo vedremo nell'esercizio svolto.

### Il lemma di Euclide

**Lemma.** Se $p$ è primo e $p\mid ab$, allora $p\mid a$ oppure $p\mid b$.

**Dimostrazione.** Supponiamo $p\nmid a$ (altrimenti abbiamo già finito). Poiché $p$ è primo, i suoi unici divisori positivi sono $1$ e $p$; non dividendo $a$, l'unico divisore comune di $p$ e $a$ è $1$, cioè $\gcd(p,a)=1$. Per Bézout, esistono $x,y$ con $px+ay=1$. Moltiplicando per $b$: $pbx+aby=b$. Ora, $p\mid pbx$ (ovviamente) e $p\mid aby$ (perché $p\mid ab$ per ipotesi, quindi $p\mid aby$); quindi $p$ divide la somma $pbx+aby=b$. $\blacksquare$

**Perché questo lemma è specifico dei numeri primi.** Non vale per numeri composti: $6\mid4\cdot3=12$, ma $6\nmid4$ e $6\nmid3$. È esattamente questa proprietà — "se divido un prodotto, divido uno dei fattori" — che rende i primi i mattoni giusti per la fattorizzazione.

### Il teorema fondamentale dell'aritmetica

**Teorema.** Ogni intero $n>1$ si scrive in modo unico (a meno dell'ordine dei fattori) come prodotto di numeri primi.

**Dimostrazione dell'esistenza (per induzione forte, notebook 10.4).** Se $n$ è primo, è già un prodotto di un solo primo (se stesso). Altrimenti $n$ è composto, cioè $n=ab$ con $1<a,b<n$; per ipotesi induttiva (forte, applicata ad $a,b<n$) sia $a$ che $b$ si scrivono come prodotti di primi, quindi anche $n=ab$.

**Dimostrazione dell'unicità.** Supponiamo $n=p_1p_2\cdots p_k=q_1q_2\cdots q_m$, due fattorizzazioni in primi. Poiché $p_1\mid n=q_1\cdots q_m$, applicando il lemma di Euclide ripetutamente (per induzione sul numero di fattori $q_i$), $p_1$ deve dividere qualche $q_i$; ma $q_i$ è primo, quindi $p_1=q_i$ (i suoi unici divisori positivi sono $1$ e $q_i$ stesso, e $p_1>1$). Riordinando, possiamo assumere $i=1$: $p_1=q_1$. Dividendo entrambi i membri per $p_1$, resta $p_2\cdots p_k=q_2\cdots q_m$, e si ripete l'argomento per induzione sul numero di fattori. Alla fine, le due fattorizzazioni coincidono a meno dell'ordine. $\blacksquare$

Questo teorema — spesso chiamato semplicemente "fattorizzazione unica" — è il motivo per cui i numeri primi meritano tutta l'attenzione che riceveranno nei prossimi notebook: sono, in un senso reso ora completamente preciso, gli **atomi moltiplicativi** di $\mathbb Z$.


```python
from math import gcd

def euclide_esteso(a, b):
    # Restituisce (d, x, y) con d = gcd(a,b) = a*x + b*y
    old_r, r = a, b
    old_x, x = 1, 0
    old_y, y = 0, 1
    while r != 0:
        q = old_r // r
        old_r, r = r, old_r - q * r
        old_x, x = x, old_x - q * x
        old_y, y = y, old_y - q * y
    return old_r, old_x, old_y

# Verifichiamo l'algoritmo di Euclide esteso e l'identita' di Bezout su alcuni esempi
for a, b in [(252, 105), (1071, 462), (17, 5), (100, 23)]:
    d, x, y = euclide_esteso(a, b)
    print(f"gcd({a},{b}) = {d}  (math.gcd: {gcd(a, b)})   Bezout: {a}*({x}) + {b}*({y}) = {a*x + b*y}")
```

## Esercizio svolto e commentato

**Problema:** calcola $\gcd(252,105)$ con l'algoritmo di Euclide, mostrando ogni passo, e trova $x,y$ con $252x+105y=\gcd(252,105)$ (identità di Bézout).

**Ragionamento:**

**Passo 1 — applichiamo ripetutamente la divisione euclidea.**
$$252=105\cdot2+42\qquad(q=2,\ r=42)$$
$$105=42\cdot2+21\qquad(q=2,\ r=21)$$
$$42=21\cdot2+0\qquad(q=2,\ r=0)$$
L'ultimo resto non nullo è $21$: per il lemma di questo notebook, $\gcd(252,105)=\gcd(105,42)=\gcd(42,21)=\gcd(21,0)=21$ (essendo $\gcd(n,0)=n$ per ogni $n>0$, ovvio dalla definizione).

**Passo 2 — risaliamo per esprimere $21$ come combinazione lineare di $252,105$ (Euclide esteso).** Dalla penultima equazione: $21=105-42\cdot2$. Dalla prima equazione, $42=252-105\cdot2$: sostituiamo
$$21=105-(252-105\cdot2)\cdot2=105-252\cdot2+105\cdot4=252\cdot(-2)+105\cdot5.$$

**Conclusione:** $x=-2,y=5$: verifichiamo $252\cdot(-2)+105\cdot5=-504+525=21$ ✓ (da verificare anche numericamente nella cella seguente, confrontando con l'algoritmo esteso già implementato).

**Perché "risalire" funziona.** Ogni riga della divisione euclidea esprime un resto come combinazione lineare intera dei due numeri della riga precedente; sostituendo a catena dal basso verso l'alto, ogni resto intermedio (incluso l'ultimo, il gcd) diventa via via una combinazione lineare intera dei due numeri originali $a,b$ — esattamente il contenuto costruttivo della dimostrazione di Bézout vista nella teoria.

## Riepilogo finale

- La **divisione euclidea** ($a=bq+r$, $0\le r<b$) esiste ed è unica; la sua dimostrazione usa il **buon ordinamento** di $\mathbb Z_{\ge0}$, uno schema dimostrativo che riappare in tutta la teoria dei numeri.
- L'**algoritmo di Euclide** calcola $\gcd(a,b)$ applicando ripetutamente $\gcd(a,b)=\gcd(b,a\bmod b)$; la sua versione "estesa" produce anche i coefficienti dell'**identità di Bézout** $ax+by=\gcd(a,b)$.
- Il **lemma di Euclide** ($p$ primo, $p\mid ab\Rightarrow p\mid a$ o $p\mid b$) è la proprietà specifica dei primi che rende possibile l'unicità della fattorizzazione.
- Il **teorema fondamentale dell'aritmetica** garantisce che ogni intero $>1$ si fattorizzi in primi in modo unico: i primi sono gli "atomi" moltiplicativi di $\mathbb Z$.
- Bézout è anche lo strumento chiave per capire, nel prossimo notebook, quando un'equazione modulare ha soluzione.
