## Perché ci interessa

L'orologio è l'esempio più familiare di "aritmetica che si ripete": $9$ ore dopo le $20{:}00$ non sono le $29{:}00$, ma le $5{:}00$ — abbiamo sottratto $24$ (o meglio, abbiamo preso il resto della divisione per $24$). Questa idea, formalizzata, è l'**aritmetica modulare**: lavorare con gli interi "dimenticando" i multipli di un certo $n$ fissato, il **modulo**.

Non è solo un gioco: l'aritmetica modulare è lo strumento con cui la teoria dei numeri diventa **calcolabile** (verificare la divisibilità di numeri enormi senza calcolarli per intero), è alla base della crittografia moderna, e — punto che ci interessa qui in modo particolare — è il primo esempio concreto e naturale di **anello quoziente**, la costruzione astratta introdotta nel notebook 11.8 senza, all'epoca, un esempio numerico così immediato.


## Teoria, passo dopo passo

### La congruenza modulo $n$

Fissato un intero $n>0$, diciamo che $a$ è **congruo** a $b$ **modulo $n$**, scritto $a\equiv b\pmod n$, se $n\mid(a-b)$ (notebook 18.5).

**La congruenza è una relazione di equivalenza** (nel senso del notebook 0.19):
- *Riflessiva:* $a\equiv a\pmod n$, perché $n\mid0=a-a$.
- *Simmetrica:* se $a\equiv b\pmod n$, cioè $n\mid(a-b)$, allora $n\mid(b-a)=-(a-b)$, quindi $b\equiv a\pmod n$.
- *Transitiva:* se $a\equiv b$ e $b\equiv c\pmod n$, cioè $n\mid(a-b)$ e $n\mid(b-c)$, allora $n\mid\big((a-b)+(b-c)\big)=(a-c)$ (chiusura per combinazioni lineari, notebook 18.5), quindi $a\equiv c\pmod n$.

Essendo una relazione di equivalenza, partiziona $\mathbb Z$ in **classi di equivalenza** (notebook 0.19): la classe di $a$, scritta $[a]_n$ o semplicemente $[a]$, è l'insieme $\{a+kn:k\in\mathbb Z\}$ di tutti gli interi congrui ad $a$. Per la divisione euclidea (notebook 18.5), ogni intero è congruo, modulo $n$, a esattamente uno tra $0,1,\ldots,n-1$ (il proprio resto nella divisione per $n$): quindi ci sono esattamente $n$ classi distinte, e l'insieme quoziente è
$$\mathbb Z/n\mathbb Z=\{[0],[1],\ldots,[n-1]\}.$$

### Le operazioni sono ben definite

Vogliamo sommare e moltiplicare classi, non singoli rappresentanti: $[a]+[b]:=[a+b]$ e $[a]\cdot[b]:=[ab]$. Il problema è che una classe ha infiniti rappresentanti (es. $[3]_5=[8]_5=[-2]_5=\ldots$): dobbiamo verificare che il risultato **non dipenda** da quale rappresentante scegliamo.

**Proposizione.** Se $a\equiv a'\pmod n$ e $b\equiv b'\pmod n$, allora $a+b\equiv a'+b'\pmod n$ e $ab\equiv a'b'\pmod n$.

**Dimostrazione.** Per ipotesi, $n\mid(a-a')$ e $n\mid(b-b')$. Per la somma: $(a+b)-(a'+b')=(a-a')+(b-b')$, somma di due multipli di $n$, quindi ancora multiplo di $n$ (chiusura per combinazioni lineari): $a+b\equiv a'+b'\pmod n$. Per il prodotto: scriviamo $a=a'+kn$, $b=b'+mn$ per opportuni $k,m\in\mathbb Z$; allora
$$ab=(a'+kn)(b'+mn)=a'b'+n(a'm+b'k+kmn),$$
quindi $ab-a'b'=n(a'm+b'k+kmn)$ è multiplo di $n$: $ab\equiv a'b'\pmod n$. $\blacksquare$

Questo è precisamente ciò che serve perché la somma e il prodotto di classi siano **ben definiti**: il risultato $[a+b]$ (o $[ab]$) è lo stesso qualunque rappresentante $a,b$ scegliamo nelle rispettive classi.

### $\mathbb Z/n\mathbb Z$ è un anello commutativo

Con queste operazioni, $\mathbb Z/n\mathbb Z$ eredita da $\mathbb Z$ tutte le proprietà di anello commutativo (notebook 11.6): associatività, commutatività, distributività, elemento neutro $[0]$ per la somma e $[1]$ per il prodotto, opposto $-[a]=[-a]$ — tutte proprietà che si verificano riportandole, tramite la ben definizione appena dimostrata, alle stesse proprietà già note in $\mathbb Z$.

Anzi, $\mathbb Z/n\mathbb Z$ è **esattamente** l'anello quoziente del notebook 11.8: l'insieme $n\mathbb Z=\{nk:k\in\mathbb Z\}$ dei multipli di $n$ è un **ideale** di $\mathbb Z$ (notebook 11.7: chiuso per somma e per moltiplicazione con qualunque elemento dell'anello — entrambe verifiche immediate), e le classi $[a]_n$ altro non sono che i laterali $a+n\mathbb Z$ dell'anello quoziente $\mathbb Z/(n)$. Il linguaggio astratto del Modulo 11 e l'aritmetica dell'orologio sono, letteralmente, lo stesso oggetto.

### Quando $\mathbb Z/n\mathbb Z$ è un campo

**Proposizione.** $[a]$ ha un inverso moltiplicativo in $\mathbb Z/n\mathbb Z$ (cioè esiste $[b]$ con $[a][b]=[1]$) se e solo se $\gcd(a,n)=1$.

**Dimostrazione.** Se $\gcd(a,n)=1$, per l'identità di Bézout (notebook 18.5) esistono $x,y$ con $ax+ny=1$. Modulo $n$, il termine $ny\equiv0$, quindi $ax\equiv1\pmod n$: $[x]$ è l'inverso cercato. Viceversa, se esiste $[b]$ con $ab\equiv1\pmod n$, cioè $ab-1=kn$ per qualche $k$, allora $ab-kn=1$: ogni divisore comune $d$ di $a,n$ divide $ab-kn=1$ (chiusura per combinazioni lineari), quindi $d\le1$, cioè $\gcd(a,n)=1$. $\blacksquare$

**Corollario.** $\mathbb Z/n\mathbb Z$ è un campo (notebook 11.10, ogni elemento non nullo invertibile) se e solo se **ogni** intero $1,\ldots,n-1$ è coprimo con $n$ — cosa che accade esattamente quando $n$ è **primo** (se $n$ fosse composto, un suo fattore proprio $d$ con $1<d<n$ avrebbe $\gcd(d,n)=d>1$, quindi $[d]$ non sarebbe invertibile).

Questo corollario è il primo grande motivo per cui i numeri primi meritano un notebook a parte, il prossimo.


```python
import numpy as np

# Tavole di addizione e moltiplicazione modulo n: costruiamo quella di mod 5 (primo)
# e mod 6 (composto), per vedere la differenza strutturale annunciata nella teoria.

def tavola_prodotto(n):
    return np.array([[(a * b) % n for b in range(n)] for a in range(n)])

for n in (5, 6):
    T = tavola_prodotto(n)
    print(f"Tavola moltiplicativa mod {n}:")
    print(T)
    invertibili = [a for a in range(1, n) if any(T[a, b] == 1 for b in range(n))]
    print(f"Elementi invertibili (coprimi con {n}): {invertibili}")
    if n != 5:
        # cerchiamo zero-divisori: a,b != 0 con a*b == 0 mod n
        zero_divisori = [(a, b) for a in range(1, n) for b in range(1, n) if T[a, b] == 0]
        print(f"Coppie (a,b) con a*b = 0 mod {n}, a,b != 0 (zero-divisori): {zero_divisori}")
    print()
```

## Esercizio svolto e commentato

**Problema:** trova l'inverso moltiplicativo di $3$ modulo $7$, cioè $[x]$ con $3x\equiv1\pmod7$.

**Ragionamento:**

**Passo 1 — verifichiamo che l'inverso esiste.** $7$ è primo, quindi $\gcd(3,7)=1$ (essendo $3<7$ e $7$ primo, l'unico divisore comune possibile è $1$): per la proposizione della teoria, l'inverso esiste.

**Passo 2 — usiamo Bézout (algoritmo di Euclide esteso, notebook 18.5) per trovare $x,y$ con $3x+7y=1$.** Applichiamo l'algoritmo di Euclide: $7=3\cdot2+1$, quindi $1=7-3\cdot2$, cioè $3\cdot(-2)+7\cdot1=1$: $x=-2$.

**Passo 3 — riduciamo $x$ modulo $7$ per ottenere un rappresentante standard.** $x=-2\equiv5\pmod7$ (essendo $-2+7=5$).

**Conclusione:** $3^{-1}\equiv5\pmod7$, cioè $3\cdot5=15\equiv1\pmod7$ (da verificare: $15=7\cdot2+1$ ✓).

**Perché funziona sempre così.** Il passo 2 è esattamente il contenuto costruttivo della dimostrazione di Bézout: l'algoritmo di Euclide esteso, applicato a $(a,n)$ con $\gcd(a,n)=1$, produce direttamente $x$ tale che $ax\equiv1\pmod n$ — non serve altro che ridurre $x$ nell'intervallo $[0,n-1]$.


```python
def euclide_esteso(a, b):
    old_r, r = a, b
    old_x, x = 1, 0
    while r != 0:
        q = old_r // r
        old_r, r = r, old_r - q * r
        old_x, x = x, old_x - q * x
    return old_r, old_x  # d = gcd(a,b), x tale che a*x + b*y = d per qualche y

n = 7
a = 3
d, x = euclide_esteso(a, n)
assert d == 1, "a e n devono essere coprimi"
inverso = x % n

print(f"gcd({a},{n}) = {d}")
print(f"x trovato dall'algoritmo esteso: {x}, ridotto mod {n}: {inverso}")
print(f"Verifica: {a}*{inverso} mod {n} = {(a * inverso) % n}")
```

## Riepilogo finale

- La **congruenza modulo $n$** ($a\equiv b\pmod n\iff n\mid(a-b)$) è una relazione di equivalenza; le sue classi formano l'insieme quoziente $\mathbb Z/n\mathbb Z=\{[0],\ldots,[n-1]\}$.
- Somma e prodotto di classi sono **ben definiti** (non dipendono dal rappresentante scelto), e rendono $\mathbb Z/n\mathbb Z$ un **anello commutativo** — anzi, esattamente l'anello quoziente $\mathbb Z/(n)$ del notebook 11.8.
- $[a]$ è invertibile in $\mathbb Z/n\mathbb Z$ se e solo se $\gcd(a,n)=1$, per Bézout (notebook 18.5); l'algoritmo di Euclide esteso calcola l'inverso esplicitamente.
- $\mathbb Z/n\mathbb Z$ è un **campo** se e solo se $n$ è **primo** (altrimenti ha divisori dello zero): il ponte diretto verso il prossimo notebook.
