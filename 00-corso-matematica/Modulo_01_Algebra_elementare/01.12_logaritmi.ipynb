## Perché ci interessa

Nel notebook precedente abbiamo lasciato in sospeso un problema: risolvere $(1.05)^t=2$, dove l'incognita è **all'esponente**. Con le equazioni di primo e secondo grado (notebook 1.5, 1.6) l'incognita era la base di un'operazione; qui è l'esponente. Ci serve un'operazione che faccia l'opposto dell'elevare a potenza: **data la base e il risultato, trovare l'esponente**.

Questa operazione si chiama **logaritmo**. È l'operazione inversa dell'esponenziale (notebook 1.11), esattamente come la sottrazione è l'inversa dell'addizione (notebook 0.2) e la radice è l'inversa dell'elevamento a potenza con esponente intero (notebook 0.6).

## Teoria, passo dopo passo

### La definizione

Fissata una base $a>0$, $a\ne1$, il **logaritmo in base $a$** di un numero $x>0$ è l'esponente a cui bisogna elevare $a$ per ottenere $x$:

$$\log_a x = y \quad \Longleftrightarrow \quad a^y=x$$

Esempio: $\log_2 8 = 3$, perché $2^3=8$. Esempio: $\log_{10} 100=2$, perché $10^2=100$.

**Perché serve $x>0$:** l'esponenziale $a^y$ è sempre positiva (notebook 1.11, codominio $(0,+\infty)$), quindi non esiste nessun $y$ tale che $a^y$ sia negativo o zero. Il logaritmo di un numero negativo o di zero **non è definito** (sui numeri reali).

### Logaritmo come funzione inversa

Se $f(x)=a^x$ è l'esponenziale, allora $f^{-1}(x)=\log_a x$ è la sua funzione inversa (il concetto generale di funzione inversa lo formalizzeremo nel Modulo 3, ma l'idea è già intuibile: "disfare" ciò che $f$ fa). Questo significa:

$$\log_a(a^x) = x \qquad \qquad a^{\log_a x} = x \ (x>0)$$

Il dominio del logaritmo è $(0,+\infty)$ (il codominio dell'esponenziale) e il suo codominio è tutto $\mathbb{R}$ (il dominio dell'esponenziale): dominio e codominio si **scambiano**, proprietà tipica delle funzioni inverse.

### Le proprietà dei logaritmi

Le proprietà dei logaritmi derivano direttamente da quelle delle potenze (notebook 1.10), tramite la definizione. Vediamo la più importante, dimostrandola:

$$\log_a(xy) = \log_a x + \log_a y$$

**Dimostrazione**: chiamiamo $p=\log_a x$ e $q=\log_a y$, cioè $x=a^p$ e $y=a^q$ (per definizione di logaritmo). Allora:

$$xy = a^p \cdot a^q = a^{p+q}$$

(abbiamo usato la proprietà delle potenze $a^p a^q=a^{p+q}$, notebook 1.10). Ma questo dice esattamente che $p+q$ è l'esponente a cui elevare $a$ per ottenere $xy$, cioè per definizione $\log_a(xy)=p+q=\log_a x+\log_a y$. ∎

Con dimostrazioni analoghe si ottengono le altre proprietà:

$$\log_a\left(\frac{x}{y}\right) = \log_a x - \log_a y \qquad \log_a(x^k) = k\log_a x$$

$$\log_a a = 1 \qquad \log_a 1 = 0$$

(le ultime due si verificano direttamente dalla definizione: $a^1=a$ e $a^0=1$).

### Cambiamento di base

A volte serve calcolare un logaritmo in una base diversa da quella disponibile (per esempio su una calcolatrice, che spesso ha solo $\log_{10}$ o $\ln=\log_e$). La formula del cambiamento di base:

$$\log_a x = \frac{\log_b x}{\log_b a}$$

per qualunque base ausiliaria $b$. Non la dimostriamo qui in dettaglio (si ottiene applicando $\log_b$ a entrambi i membri di $a^{\log_a x}=x$ e usando $\log_b(x^k)=k\log_b x$), ma è uno strumento pratico da ricordare.

### I logaritmi "standard"

Due basi hanno notazioni dedicate: $\log_{10}x$ si scrive spesso semplicemente $\log x$ (**logaritmo decimale**); $\log_e x$ si scrive $\ln x$ (**logaritmo naturale**), e sarà la base che useremo quasi sempre a partire dal Modulo 4, per la sua relazione speciale con la derivata (Modulo 5).

````python
import sympy as sp

# Verifica della definizione: log_2(8) = 3 perché 2^3=8
print("log_2(8) =", sp.log(8, 2))

# Verifica della proprietà del prodotto, dimostrata sopra
x, y, a = sp.symbols('x y a', positive=True)
lato_sinistro = sp.log(x*y, a)
lato_destro = sp.log(x, a) + sp.log(y, a)
print("\nlog_a(xy) - (log_a(x)+log_a(y)) semplificato:", sp.simplify(lato_sinistro - lato_destro))

# Verifica numerica delle proprietà
import math
print("\nlog_2(8*4) =", math.log2(8*4))
print("log_2(8) + log_2(4) =", math.log2(8) + math.log2(4))

# Cambiamento di base: log_2(10) calcolato via log naturale
log2_10_diretto = math.log2(10)
log2_10_cambiamento = math.log(10) / math.log(2)  # ln(10)/ln(2)
print(f"\nlog_2(10) diretto: {log2_10_diretto:.6f}")
print(f"log_2(10) via cambiamento di base: {log2_10_cambiamento:.6f}")
````

## Esercizio svolto e commentato

**Problema:** risolvi $(1.05)^t=2$ (il problema lasciato in sospeso nel notebook 1.11), trovando il valore esatto di $t$ come espressione con i logaritmi, poi approssimalo numericamente.

**Ragionamento:**

**Passo 1 — applichiamo il logaritmo a entrambi i membri.** Possiamo farlo perché, se due quantità positive sono uguali, i loro logaritmi (in una base qualsiasi, scegliamo $10$) sono anch'essi uguali — è l'invariantiva (notebook 0.5) applicata alla funzione logaritmo invece che a un numero:

$$\log\big((1.05)^t\big) = \log 2$$

**Passo 2 — usiamo la proprietà del logaritmo di una potenza**, $\log_a(x^k)=k\log_a x$, per "tirare giù" l'esponente $t$:

$$t\cdot\log(1.05) = \log 2$$

**Passo 3 — isoliamo $t$** dividendo per $\log(1.05)$ (lecito, perché $\log(1.05)\ne0$: infatti $\log_a x=0$ solo se $x=1$, e qui $1.05\ne1$):

$$t = \frac{\log 2}{\log(1.05)}$$

Questa è la soluzione **esatta**, scritta come rapporto di logaritmi. Calcoliamo il valore numerico: $\log2\approx0.30103$, $\log(1.05)\approx0.02119$, quindi:

$$t \approx \frac{0.30103}{0.02119} \approx 14.2$$

Coerente con l'approssimazione per tentativi del notebook precedente (tra 14 e 15 anni).

**Perché "tirare giù" l'esponente con il logaritmo funziona sempre, a differenza del tentativi numerici:** il metodo per tentativi (provare $t=10,14,15,\dots$) dà solo un'approssimazione, per quanto buona. Il logaritmo fornisce invece il valore **esatto** (come espressione), che poi possiamo approssimare numericamente con la precisione che vogliamo — è la differenza tra "indovinare per tentativi" e "risolvere con uno strumento apposito".

````python
import sympy as sp

t = sp.symbols('t', positive=True)
soluzione_esatta = sp.solve(sp.Eq((sp.Rational(105, 100))**t, 2), t)
print("Soluzione esatta (forma simbolica):", soluzione_esatta)
print("Valore numerico:", float(soluzione_esatta[0]))
````

## Riepilogo finale

- $\log_a x$ (con $a>0,a\ne1,x>0$) è l'esponente a cui elevare $a$ per ottenere $x$: è la funzione inversa dell'esponenziale $a^x$ (notebook 1.11).
- Proprietà principali, dimostrate a partire dalla definizione e dalle proprietà delle potenze: $\log_a(xy)=\log_ax+\log_ay$, $\log_a(x/y)=\log_ax-\log_ay$, $\log_a(x^k)=k\log_ax$.
- Il logaritmo permette di "tirare giù" un esponente incognito, risolvendo equazioni come $a^t=k$ che l'invariantiva da sola non può risolvere.
- $\log x$ (senza base, sottintesa 10) è il logaritmo decimale; $\ln x$ è il logaritmo naturale (base $e$), che useremo sempre di più da qui in avanti.
