# 1.6 — Equazioni di secondo grado

**Prerequisiti:** 1.2 Prodotti notevoli, 1.3 Scomposizione dei polinomi, 1.5 Equazioni di primo grado, 0.6 Frazioni, potenze e radici.


## Perché ci interessa

Con le equazioni di primo grado, isolare $x$ con l'invariantiva bastava sempre. Ma cosa succede con $x^2-5x+6=0$? Qui $x$ compare due volte, con esponenti diversi: non possiamo più "raccogliere" un singolo termine con $x$, perché $x^2$ e $x$ non sono simili (notebook 1.1).

Ci serve una strategia nuova. Vedremo che ce ne sono in realtà due complementari: riconoscere una scomposizione (quando possibile) oppure usare una formula generale, valida sempre.

## Teoria, passo dopo passo

### La forma generale

Un'equazione di secondo grado si scrive sempre, dopo aver spostato tutto da un lato (invariantiva, notebook 0.5), come:

$$ax^2+bx+c=0 \qquad (a\ne0)$$

Se $a=0$, l'equazione degenera in una di primo grado (notebook 1.5): per questo richiediamo $a\ne0$.

### Strategia 1: risolvere per scomposizione, quando è comoda

Se il trinomio $ax^2+bx+c$ si scompone facilmente (notebook 1.3), possiamo sfruttare la **legge di annullamento del prodotto**: un prodotto di fattori è zero se e solo se almeno uno dei fattori è zero.

Esempio: $x^2-5x+6=0$. Scomponiamo (cerchiamo due numeri che sommati diano $-5$ e moltiplicati diano $6$: sono $-2$ e $-3$):

$$(x-2)(x-3)=0$$

Per la legge di annullamento del prodotto: $x-2=0$ oppure $x-3=0$, quindi $x=2$ oppure $x=3$.

**Perché questa legge è vera:** se $x-2\ne0$ e $x-3\ne0$, allora entrambi i fattori sono numeri reali diversi da zero, e il prodotto di due numeri diversi da zero non può mai essere zero (in $\mathbb{R}$ non ci sono "divisori dello zero": è una conseguenza della struttura dei numeri reali che diamo per buona qui, la riprenderemo con più rigore nel Modulo 11).

### Strategia 2: la formula risolutiva generale

Non tutti i trinomi si scompongono a vista. Serve un metodo che funzioni **sempre**. Partiamo da $ax^2+bx+c=0$ e completiamo il quadrato (una tecnica che userai ancora, per esempio nel Modulo 2 per le coniche): moltiplichiamo tutto per $4a$ (invariantiva, $a\ne0$) e riorganizziamo per far comparire il quadrato di un binomio (notebook 1.2):

$$4a^2x^2+4abx+4ac=0 \quad\Rightarrow\quad (2ax+b)^2 - b^2+4ac=0 \quad\Rightarrow\quad (2ax+b)^2 = b^2-4ac$$

Chiamiamo $\Delta = b^2-4ac$ il **discriminante**. Se $\Delta\ge0$, possiamo estrarre la radice quadrata di entrambi i membri (notebook 0.6, ricordando che la radice di un quadrato è un valore assoluto — qui gestito considerando entrambi i segni):

$$2ax+b = \pm\sqrt{\Delta} \quad\Rightarrow\quad x = \frac{-b\pm\sqrt{\Delta}}{2a}$$

Questa è la **formula risolutiva** (o "formula quadratica"), valida per ogni equazione di secondo grado.

### I tre casi del discriminante

Il segno di $\Delta=b^2-4ac$ determina quante soluzioni reali esistono:

- $\Delta>0$: due soluzioni reali **distinte**, $x_{1,2}=\dfrac{-b\pm\sqrt{\Delta}}{2a}$.
- $\Delta=0$: una soluzione reale, detta **doppia** (o di molteplicità 2), $x=-\dfrac{b}{2a}$ (i due "rami" $\pm\sqrt{\Delta}$ coincidono perché $\sqrt{0}=0$).
- $\Delta<0$: **nessuna soluzione reale**, perché $\sqrt{\Delta}$ richiederebbe la radice di un numero negativo (la troveremo nel Modulo 9, con i numeri complessi).

### Un caso comodo: $b=0$

Se $b=0$, l'equazione diventa $ax^2+c=0$, risolvibile direttamente senza formula: $x^2=-\frac{c}{a}$, quindi $x=\pm\sqrt{-\frac{c}{a}}$ (se $-\frac{c}{a}\ge0$, altrimenti nessuna soluzione reale).

````python
import sympy as sp

x, a, b, c = sp.symbols('x a b c')

# Verifica simbolica della formula risolutiva (completamento del quadrato)
delta = b**2 - 4*a*c
x1 = (-b + sp.sqrt(delta)) / (2*a)
x2 = (-b - sp.sqrt(delta)) / (2*a)

# Sostituendo x1 nell'equazione originale, il risultato deve semplificarsi a 0
verifica = sp.simplify(a*x1**2 + b*x1 + c)
print("a*x1^2+b*x1+c semplificato:", verifica)

# Esempio numerico: x^2-5x+6=0
soluzioni_scomposizione = sp.solve(sp.factor(x**2 - 5*x + 6), x)
print("\nSoluzioni di x^2-5x+6=0 (via scomposizione):", soluzioni_scomposizione)

# I tre casi del discriminante
casi = [(1, -5, 6), (1, -4, 4), (1, 0, 1)]  # delta>0, delta=0, delta<0
for a_val, b_val, c_val in casi:
    delta_val = b_val**2 - 4*a_val*c_val
    soluzioni = sp.solve(sp.Eq(a_val*x**2 + b_val*x + c_val, 0), x)
    print(f"\na={a_val}, b={b_val}, c={c_val}: delta={delta_val}, soluzioni={soluzioni}")
````

## Esercizio svolto e commentato

**Problema:** risolvi $2x^2-3x-2=0$ con la formula risolutiva, verificando poi con la scomposizione.

**Ragionamento:**

**Passo 1 — identifichiamo i coefficienti**: $a=2$, $b=-3$, $c=-2$.

**Passo 2 — calcoliamo il discriminante**:

$$\Delta = b^2-4ac = (-3)^2-4(2)(-2) = 9+16=25$$

Poiché $\Delta=25>0$ (un quadrato perfetto, comodo), ci aspettiamo due soluzioni reali distinte, esprimibili senza radici "scomode".

**Passo 3 — applichiamo la formula**:

$$x = \frac{-b\pm\sqrt{\Delta}}{2a} = \frac{3\pm\sqrt{25}}{4} = \frac{3\pm5}{4}$$

Separiamo i due casi:

$$x_1 = \frac{3+5}{4} = \frac{8}{4} = 2 \qquad x_2 = \frac{3-5}{4} = \frac{-2}{4} = -\frac{1}{2}$$

**Verifica con la scomposizione**: se $x_1=2$ e $x_2=-\frac{1}{2}$ sono le radici, il polinomio si scompone come $a(x-x_1)(x-x_2)$ (una proprietà che vedremo meglio nel Modulo 3, parlando di funzioni polinomiali):

$$2(x-2)\left(x+\frac{1}{2}\right) = 2x^2-3x-2$$

Sviluppando il lato sinistro con la distributiva, si può verificare che coincide con l'equazione di partenza. ✓

**Perché il discriminante essere un quadrato perfetto (qui $25=5^2$) è una fortuna, non una regola generale:** quando $\Delta$ non è un quadrato perfetto (es. $\Delta=7$), le soluzioni contengono $\sqrt{7}$ e restano scritte con la radice — non c'è nulla di sbagliato, sono comunque soluzioni valide ed esatte, solo meno "pulite" da leggere.

````python
import sympy as sp

x = sp.symbols('x')
a_val, b_val, c_val = 2, -3, -2
delta = b_val**2 - 4*a_val*c_val
print("Discriminante:", delta)

soluzioni = sp.solve(sp.Eq(a_val*x**2 + b_val*x + c_val, 0), x)
print("Soluzioni:", soluzioni)

# Verifica tramite scomposizione
scomposizione = sp.factor(a_val*x**2 + b_val*x + c_val)
print("Scomposizione:", scomposizione)
````

## Riepilogo finale

- Un'equazione di secondo grado ha la forma $ax^2+bx+c=0$ ($a\ne0$); si può risolvere per **scomposizione** (legge di annullamento del prodotto) quando è comoda, oppure sempre con la **formula risolutiva** $x=\dfrac{-b\pm\sqrt{\Delta}}{2a}$, dove $\Delta=b^2-4ac$.
- Il segno del **discriminante** $\Delta$ determina il numero di soluzioni reali: $\Delta>0$ due distinte, $\Delta=0$ una doppia, $\Delta<0$ nessuna.
- La legge di annullamento del prodotto (usata nella scomposizione) dice che un prodotto è zero solo se almeno un fattore è zero.
- Se $b=0$, conviene risolvere direttamente come $x^2=-\frac{c}{a}$, senza la formula generale.
