## Perché ci interessa

Nei due notebook precedenti abbiamo introdotto esponenziali e logaritmi, e già risolto un'equazione esponenziale ($( 1.05)^t=2$) come esempio applicativo. Ora sistematizziamo: quali sono le strategie generali per risolvere equazioni dove l'incognita compare come esponente, o dentro un logaritmo?

Vedremo che, come per le equazioni algebriche (notebook 1.5, 1.6), esiste un repertorio di tecniche riconoscibili, da scegliere a seconda della forma dell'equazione.

## Teoria, passo dopo passo

### Equazioni esponenziali con basi uguali (o riconducibili)

Se un'equazione si scrive come $a^{f(x)}=a^{g(x)}$ (stessa base su entrambi i lati), possiamo sfruttare il fatto che $a^x$ è una funzione **iniettiva** (per $a>0,a\ne1$: essendo strettamente crescente o decrescente, notebook 1.11, non assume mai due volte lo stesso valore — la stessa idea usata nell'esercizio del notebook 1.10 per $a^x=1$). Quindi:

$$a^{f(x)}=a^{g(x)} \quad \Longleftrightarrow \quad f(x)=g(x)$$

Esempio: $2^{x+1}=2^{5}$. Poiché le basi coincidono, uguagliamo gli esponenti: $x+1=5 \Rightarrow x=4$.

A volte le basi non sono uguali "a vista" ma lo diventano dopo aver riscritto i numeri come potenze di una base comune: $4^x=8$ si riscrive come $(2^2)^x=2^3$, cioè $2^{2x}=2^3$, da cui $2x=3 \Rightarrow x=\frac32$.

### Equazioni esponenziali con basi diverse: il logaritmo

Quando non è possibile ricondursi a una base comune (come nell'esempio $(1.05)^t=2$ del notebook 1.12), si applica il logaritmo a entrambi i membri — la tecnica generale vista nell'esercizio svolto del notebook 1.12:

$$a^{f(x)}=b \quad \Rightarrow \quad f(x)\log a=\log b \quad \Rightarrow \quad f(x)=\frac{\log b}{\log a}$$

### Sostituzione: quando l'esponenziale compare "elevata"

A volte l'equazione contiene $a^x$ elevato a sua volta, per esempio $a^{2x}-3a^x+2=0$. Notiamo che $a^{2x}=(a^x)^2$ (proprietà delle potenze, notebook 1.10): possiamo allora **sostituire** $t=a^x$, trasformando l'equazione in una di secondo grado nell'incognita ausiliaria $t$ (notebook 1.6):

$$t^2-3t+2=0 \quad\Rightarrow\quad (t-1)(t-2)=0 \quad\Rightarrow\quad t=1 \text{ oppure } t=2$$

Poi si torna alla $x$ originale, risolvendo $a^x=1$ e $a^x=2$ separatamente. **Attenzione**: poiché $a^x>0$ sempre (notebook 1.11), eventuali soluzioni $t\le0$ vanno scartate — non esiste nessun $x$ per cui $a^x$ sia negativo o zero.

### Equazioni logaritmiche

Se l'equazione si scrive come $\log_a f(x) = \log_a g(x)$ (stessa base), il logaritmo è anch'esso iniettivo (essendo l'inversa di una funzione iniettiva), quindi:

$$\log_a f(x)=\log_a g(x) \quad \Longleftrightarrow \quad f(x)=g(x) \quad \text{con le condizioni } f(x)>0,\ g(x)>0$$

**Le condizioni di esistenza sono fondamentali**: il logaritmo è definito solo per argomenti positivi (notebook 1.12). Risolvere l'equazione senza controllare le C.E. può produrre soluzioni "false", da scartare.

Esempio: $\log_2(x-1) = \log_2(5-x)$. Condizioni di esistenza: $x-1>0$ e $5-x>0$, cioè $1<x<5$. Uguagliando gli argomenti: $x-1=5-x \Rightarrow 2x=6 \Rightarrow x=3$. Verifichiamo che $x=3$ rispetti le C.E.: $1<3<5$. ✓ Soluzione accettabile.

Se l'equazione ha un solo logaritmo, tipo $\log_2(x+3)=4$, si applica direttamente la definizione (notebook 1.12): $x+3=2^4=16 \Rightarrow x=13$ (verificando che $x+3=16>0$, C.E. rispettata).

````python
import sympy as sp

x, t = sp.symbols('x t')

# Equazione con basi uguali
sol1 = sp.solve(sp.Eq(2*x + 2, 10), x)  # da 2^(x+1)=2^5 -> x+1=5
print("Esempio basi uguali (2^(x+1)=2^5):", sol1)

# Equazione con sostituzione: a^(2x) - 3a^x + 2 = 0, con a=2
equazione_sostituita = sp.Eq(t**2 - 3*t + 2, 0)
soluzioni_t = sp.solve(equazione_sostituita, t)
print("\nSoluzioni per t:", soluzioni_t)
for t_val in soluzioni_t:
    sol_x = sp.solve(sp.Eq(2**x, t_val), x)
    print(f"  2^x = {t_val}  ->  x = {sol_x}")

# Equazione logaritmica con basi uguali
sol_log = sp.solve(sp.Eq(sp.log(x - 1, 2), sp.log(5 - x, 2)), x)
print("\nlog_2(x-1)=log_2(5-x):", sol_log)

# Verifica C.E.
for candidato in sol_log:
    ce_ok = (candidato - 1 > 0) and (5 - candidato > 0)
    print(f"  x={candidato}: rispetta le C.E. (1<x<5)? {ce_ok}")
````

## Esercizio svolto e commentato

**Problema:** risolvi $4^x - 5\cdot2^x + 4=0$.

**Ragionamento:**

**Passo 1 — riconosciamo che $4^x$ si può riscrivere in funzione di $2^x$.** Poiché $4=2^2$, abbiamo $4^x=(2^2)^x=2^{2x}=(2^x)^2$ (proprietà delle potenze, notebook 1.10). L'equazione diventa:

$$(2^x)^2 - 5\cdot2^x+4=0$$

**Passo 2 — sostituiamo $t=2^x$** (ricordando che dovrà risultare $t>0$, perché $2^x$ è sempre positivo, notebook 1.11):

$$t^2-5t+4=0$$

**Passo 3 — risolviamo l'equazione di secondo grado in $t$** (notebook 1.6): cerchiamo due numeri che sommati diano $5$ e moltiplicati diano $4$: sono $1$ e $4$.

$$(t-1)(t-4)=0 \quad\Rightarrow\quad t=1 \text{ oppure } t=4$$

Entrambi i valori sono positivi: nessuno va scartato.

**Passo 4 — torniamo alla $x$ originale**, risolvendo $2^x=t$ per ciascun valore trovato:

- $2^x=1$: poiché $2^0=1$, otteniamo $x=0$.
- $2^x=4$: poiché $2^2=4$, otteniamo $x=2$.

Le soluzioni sono $x=0$ e $x=2$.

**Verifica** (sostituendo nell'equazione originale): per $x=0$: $4^0-5\cdot2^0+4 = 1-5+4=0$. ✓ Per $x=2$: $4^2-5\cdot2^2+4=16-20+4=0$. ✓

**Perché controllare il segno di $t$ è un passaggio che non si può saltare:** se l'equazione di secondo grado in $t$ avesse prodotto una soluzione negativa (per esempio $t=-3$), quella soluzione andrebbe **scartata subito**, senza nemmeno provare a risolvere $2^x=-3$: sappiamo già, dalla teoria (notebook 1.11), che $2^x$ non può mai essere negativo, quindi quell'equazione non avrebbe soluzione.

````python
import sympy as sp

x = sp.symbols('x')
soluzioni = sp.solve(sp.Eq(4**x - 5*2**x + 4, 0), x)
print("Soluzioni:", soluzioni)
````

## Riepilogo finale

- Se le basi coincidono (o si possono ricondurre a una base comune), $a^{f(x)}=a^{g(x)} \iff f(x)=g(x)$, sfruttando l'iniettività dell'esponenziale.
- Con basi diverse, si applica il logaritmo a entrambi i membri per "tirare giù" l'esponente incognito.
- Se compare $a^{2x}$ insieme ad $a^x$, la sostituzione $t=a^x$ riconduce a un'equazione algebrica nota (spesso di secondo grado) — ricordando sempre di scartare le soluzioni $t\le0$.
- Per le equazioni logaritmiche, oltre a risolvere $f(x)=g(x)$, bisogna sempre verificare le **condizioni di esistenza** (argomenti positivi) prima di accettare una soluzione.
