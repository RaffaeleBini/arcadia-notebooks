# 1.9 — Disequazioni intere e fratte

**Prerequisiti:** 1.6 Equazioni di secondo grado, 1.8 Disequazioni, 1.4 Frazioni algebriche.


## Perché ci interessa

Nel notebook 1.8 abbiamo risolto disequazioni di primo grado con l'invariantiva. Ma $x^2-5x+6>0$ non si risolve così: non possiamo "isolare $x$" perché compare con due esponenti diversi (lo stesso problema che avevamo con le equazioni di secondo grado, notebook 1.6).

La soluzione è uno strumento nuovo e molto potente, lo **studio del segno**: invece di isolare $x$, studiamo *dove* un'espressione è positiva, negativa o nulla. Questo metodo funzionerà anche per le disequazioni con frazioni algebriche (notebook 1.4), dove l'incognita compare al denominatore.

## Teoria, passo dopo passo

### Disequazioni di secondo grado: il segno di una parabola

Consideriamo $x^2-5x+6>0$. Scomponiamo il trinomio (notebook 1.3, o notebook 1.6): $x^2-5x+6=(x-2)(x-3)$. La disequazione diventa:

$$(x-2)(x-3)>0$$

Ci chiediamo: quando il prodotto di due fattori è **positivo**? Per la regola dei segni della moltiplicazione (notebook 0.2): quando i due fattori hanno **lo stesso segno** (entrambi positivi, o entrambi negativi).

- $x-2>0$ e $x-3>0$: cioè $x>2$ **e** $x>3$, cioè $x>3$ (la condizione più restrittiva).
- $x-2<0$ e $x-3<0$: cioè $x<2$ **e** $x<3$, cioè $x<2$.

La soluzione è l'**unione** (notebook 0.13) di questi due casi (perché ci basta che si verifichi *uno* dei due, non entrambi contemporaneamente): $x<2$ oppure $x>3$, cioè $(-\infty,2)\cup(3,+\infty)$.

### Il metodo grafico dello studio del segno (più veloce)

C'è un modo più rapido, che userai sempre: disegnare uno **schema dei segni**. Si segnano sulla retta numerica gli **zeri** di ciascun fattore (qui $x=2$ e $x=3$), e si indica il segno di ciascun fattore in ogni intervallo che questi zeri creano. Il segno del prodotto in ciascuna zona è il prodotto dei segni dei singoli fattori (regola dei segni, notebook 0.2):

| Intervallo | $x-2$ | $x-3$ | $(x-2)(x-3)$ |
|---|---|---|---|
| $x<2$ | $-$ | $-$ | $+$ |
| $2<x<3$ | $+$ | $-$ | $-$ |
| $x>3$ | $+$ | $+$ | $+$ |

Vogliamo dove il prodotto è **positivo** ($>0$): le prime e ultime colonne, cioè $x<2$ oppure $x>3$ — coincide con quanto trovato sopra, ma il metodo dello schema è più rapido da applicare, specialmente con più di due fattori.

### Disequazioni fratte

Con una frazione algebrica (notebook 1.4), tipo $\dfrac{x-1}{x+2}\ge0$, si applica **lo stesso identico schema dei segni**, con un'unica accortezza: il denominatore **non può mai essere zero** (notebook 1.4), quindi il suo zero va sempre **escluso** dalla soluzione, anche se la disuguaglianza è "$\ge$" (mai "$\le$" o "$\ge$" incluso per un valore che annulla il denominatore).

| Intervallo | $x-1$ | $x+2$ | $\dfrac{x-1}{x+2}$ |
|---|---|---|---|
| $x<-2$ | $-$ | $-$ | $+$ |
| $-2<x<1$ | $-$ | $+$ | $-$ |
| $x>1$ | $+$ | $+$ | $+$ |

Vogliamo $\ge0$ (positivo o nullo): $x<-2$ oppure $x\ge1$ (incluso $x=1$, dove il numeratore si annulla — la frazione vale $0$, ammesso). Ma $x=-2$ resta **sempre escluso** (denominatore zero): soluzione $(-\infty,-2)\cup[1,+\infty)$.

### La strategia generale

Per qualunque disequazione (con più fattori, o con frazioni), il procedimento è sempre lo stesso:

1. Porta tutto da un lato, con $0$ dall'altro (invariantiva, notebook 0.5).
2. Scomponi numeratore e denominatore (notebook 1.3).
3. Trova gli zeri di ogni fattore.
4. Costruisci lo schema dei segni.
5. Leggi la riga del segno totale, ricordando di escludere sempre gli zeri del denominatore.

````python
import sympy as sp

x = sp.symbols('x')

# Disequazione di secondo grado
soluzione1 = sp.solve_univariate_inequality(x**2 - 5*x + 6 > 0, x)
print("x^2-5x+6>0:", soluzione1)

# Disequazione fratta
soluzione2 = sp.solve_univariate_inequality((x - 1)/(x + 2) >= 0, x)
print("(x-1)/(x+2)>=0:", soluzione2)

# Costruiamo lo schema dei segni "a mano" campionando alcuni punti in ogni zona
def segno(espressione, punto):
    valore = espressione.subs(x, punto)
    return "+" if valore > 0 else ("-" if valore < 0 else "0")

espressione = (x - 1) / (x + 2)
punti_di_prova = {"x<-2": -5, "-2<x<1": 0, "x>1": 5}
print("\nSchema dei segni di (x-1)/(x+2):")
for zona, punto in punti_di_prova.items():
    print(f"  {zona} (es. x={punto}): segno {segno(espressione, punto)}")
````

## Esercizio svolto e commentato

**Problema:** risolvi $\dfrac{x^2-4}{x-1} \le 0$.

**Ragionamento:**

**Passo 1 — l'espressione è già nella forma "$\le0$"**, con tutto da un lato: possiamo procedere direttamente allo studio del segno.

**Passo 2 — scomponiamo numeratore e denominatore.** Numeratore: $x^2-4=(x+2)(x-2)$ (differenza di quadrati, notebook 1.2). Denominatore: $x-1$, già scomposto (di primo grado, non si scompone oltre).

**Passo 3 — troviamo gli zeri di ogni fattore:** $x+2=0 \Rightarrow x=-2$; $x-2=0 \Rightarrow x=2$; $x-1=0 \Rightarrow x=1$ (**questo è il denominatore: da escludere sempre**).

**Passo 4 — costruiamo lo schema dei segni**, ordinando gli zeri sulla retta: $-2 < 1 < 2$.

| Intervallo | $x+2$ | $x-2$ | $x-1$ | $\dfrac{(x+2)(x-2)}{x-1}$ |
|---|---|---|---|---|
| $x<-2$ | $-$ | $-$ | $-$ | $-$ |
| $-2<x<1$ | $+$ | $-$ | $-$ | $+$ |
| $1<x<2$ | $+$ | $-$ | $+$ | $-$ |
| $x>2$ | $+$ | $+$ | $+$ | $+$ |

(il segno totale è il prodotto dei tre segni: due fattori negativi e uno negativo danno... contiamo i segni "$-$": nella prima riga sono 3, dispari, quindi il prodotto è negativo; nella seconda riga sono 2, pari, prodotto positivo; e così via — regola generale, notebook 0.2)

**Passo 5 — leggiamo dove vale $\le0$** (negativo o nullo): la prima riga ($x<-2$) e la terza ($1<x<2$). Includiamo gli zeri del **numeratore** (dove l'espressione vale esattamente $0$, ammesso perché la disuguaglianza è $\le$): $x=-2$ e $x=2$. **Escludiamo sempre** lo zero del denominatore, $x=1$.

$$[-2,1) \cup [2,2] = [-2,1)\cup\{2\}$$

Nota che $x=2$ da solo (il numeratore si annulla lì, quindi l'espressione vale $0\le0$, ammesso) va incluso come punto isolato, mentre l'intervallo $1<x<2$ (dove l'espressione è negativa) non arriva fino a includere altri punti oltre i suoi estremi.

**Perché $x=1$ resta escluso anche se la disequazione ammette l'uguaglianza ($\le$):** l'espressione $\frac{(x+2)(x-2)}{x-1}$ semplicemente **non è definita** in $x=1$ (notebook 1.4): non ha senso chiedersi se sia "$\le0$" o meno in un punto dove non esiste.

````python
import sympy as sp

x = sp.symbols('x')
espressione = (x**2 - 4) / (x - 1)
soluzione = sp.solve_univariate_inequality(espressione <= 0, x)
print("Soluzione:", soluzione)
````

## Riepilogo finale

- Per disequazioni con l'incognita di grado superiore al primo, o con frazioni algebriche, si usa lo **studio del segno**: scomponi, trova gli zeri di ogni fattore, costruisci uno schema dei segni.
- Il segno del prodotto/quoziente in ogni zona si ottiene contando i segni negativi dei fattori (pari = positivo, dispari = negativo), regola dei segni del notebook 0.2.
- Gli zeri del **numeratore** si includono se la disuguaglianza ammette l'uguaglianza ($\le$ o $\ge$); gli zeri del **denominatore** vanno **sempre esclusi**, indipendentemente dal verso della disuguaglianza.
- Lo schema dei segni è il metodo generale che funziona per qualunque numero di fattori.
