# 1.7 — Sistemi di equazioni

**Prerequisiti:** 1.5 Equazioni di primo grado.


## Perché ci interessa

Finora abbiamo risolto equazioni con **una** incognita. Ma molti problemi reali coinvolgono più quantità sconosciute legate da più condizioni contemporaneamente: "la somma di due numeri è 10, e la loro differenza è 2 — quali sono?". Qui servono **due** incognite e **due** equazioni, da soddisfare **insieme**.

Un insieme di equazioni che devono valere simultaneamente si chiama **sistema**. Imparare a risolverlo è il passo naturale dopo aver imparato a risolvere una singola equazione.

## Teoria, passo dopo passo

### Cos'è un sistema di equazioni lineari

Un sistema di due equazioni di primo grado (notebook 1.5) in due incognite $x,y$ si scrive:

$$\begin{cases} a_1x+b_1y=c_1 \\ a_2x+b_2y=c_2 \end{cases}$$

Una **soluzione** del sistema è una coppia $(x,y)$ che soddisfa **entrambe** le equazioni contemporaneamente — non basta soddisfarne una sola. Geometricamente (lo vedremo con precisione nel notebook 2.2), ogni equazione rappresenta una retta nel piano, e la soluzione del sistema è il punto dove le due rette si incontrano.

### Metodo di sostituzione

L'idea: ricavare un'incognita da un'equazione, e **sostituirla** nell'altra, riducendo il problema a un'equazione in una sola incognita (che sappiamo già risolvere, notebook 1.5).

Esempio: $\begin{cases} x+y=10 \\ x-y=2 \end{cases}$

**Passo 1** — dalla prima equazione ricaviamo $x$: $x=10-y$.

**Passo 2** — sostituiamo questa espressione al posto di $x$ nella seconda equazione:

$$(10-y)-y=2 \quad\Rightarrow\quad 10-2y=2 \quad\Rightarrow\quad -2y=-8 \quad\Rightarrow\quad y=4$$

**Passo 3** — sostituiamo $y=4$ nell'espressione trovata per $x$: $x=10-4=6$.

La soluzione è $(x,y)=(6,4)$.

### Metodo di somma e sottrazione (o "eliminazione")

L'idea: sommare (o sottrarre) le due equazioni, membro a membro, in modo da **far sparire** una delle incognite. Funziona subito se un'incognita ha coefficienti opposti (o uguali) nelle due equazioni.

Stesso esempio: $\begin{cases} x+y=10 \\ x-y=2 \end{cases}$. Il coefficiente di $y$ è $+1$ nella prima e $-1$ nella seconda: sommando le due equazioni membro a membro, $y$ si cancella:

$$(x+y)+(x-y) = 10+2 \quad\Rightarrow\quad 2x=12 \quad\Rightarrow\quad x=6$$

Poi sostituiamo $x=6$ in una delle due equazioni originali per trovare $y$: $6+y=10 \Rightarrow y=4$. Stessa soluzione di prima: $(6,4)$ — **come deve essere**, perché stiamo risolvendo lo stesso sistema con due strade diverse.

**Perché sommare membro a membro è lecito:** se $A=B$ e $C=D$ (le due equazioni), allora $A+C=B+D$ è ancora un'affermazione vera — è la proprietà invariantiva dell'addizione (notebook 0.5), applicata non a un numero fisso ma a un'intera equazione.

Se i coefficienti non sono già opposti, si moltiplicano le equazioni per opportuni numeri (invariantiva della moltiplicazione) prima di sommarle, in modo da renderli opposti.

### Sistemi impossibili e indeterminati

Come per le equazioni singole (notebook 1.5), un sistema può non avere soluzione (**impossibile**, le due rette sono parallele e non si incontrano mai) oppure averne infinite (**indeterminato**, le due equazioni rappresentano in realtà la stessa retta). Questi casi si riconoscono quando, applicando i metodi sopra, si arriva a un'affermazione falsa (es. $0=5$) o sempre vera (es. $0=0$) — esattamente come nel notebook 1.5.

````python
import sympy as sp

x, y = sp.symbols('x y')

# Risolviamo il sistema con sympy
sistema = [sp.Eq(x + y, 10), sp.Eq(x - y, 2)]
soluzione = sp.solve(sistema, [x, y])
print("Soluzione del sistema:", soluzione)

# Verifica manuale del metodo di sostituzione
x_espresso = 10 - y  # dalla prima equazione
equazione_sostituita = sp.Eq((10 - y) - y, 2)
y_trovato = sp.solve(equazione_sostituita, y)[0]
x_trovato = 10 - y_trovato
print(f"\nMetodo di sostituzione: x={x_trovato}, y={y_trovato}")

# Esempio di sistema impossibile: rette parallele
sistema_impossibile = [sp.Eq(x + y, 5), sp.Eq(x + y, 8)]
print("\nSistema impossibile:", sp.solve(sistema_impossibile, [x, y]))  # lista vuota

# Esempio di sistema indeterminato: stessa retta scritta due volte
sistema_indeterminato = [sp.Eq(x + y, 5), sp.Eq(2*x + 2*y, 10)]
print("Sistema indeterminato:", sp.solve(sistema_indeterminato, [x, y]))
````

## Esercizio svolto e commentato

**Problema:** risolvi $\begin{cases} 2x+3y=7 \\ 3x-y=5 \end{cases}$ con il metodo di eliminazione.

**Ragionamento:**

Nessuna incognita ha già coefficienti opposti. Scegliamo di eliminare $y$: nella prima equazione il coefficiente di $y$ è $3$, nella seconda è $-1$. Moltiplichiamo la **seconda** equazione per $3$, così il coefficiente di $y$ diventa $-3$, opposto al $3$ della prima (invariantiva della moltiplicazione, notebook 0.5 — lecito perché moltiplichiamo per $3\ne0$):

$$\begin{cases} 2x+3y=7 \\ 9x-3y=15 \end{cases}$$

**Passo 1 — sommiamo le due equazioni membro a membro**, così $y$ si elimina:

$$(2x+3y)+(9x-3y) = 7+15 \quad\Rightarrow\quad 11x=22 \quad\Rightarrow\quad x=2$$

**Passo 2 — sostituiamo $x=2$ in una delle equazioni originali** (scegliamo la seconda, perché ha coefficienti più semplici):

$$3(2)-y=5 \quad\Rightarrow\quad 6-y=5 \quad\Rightarrow\quad y=1$$

La soluzione è $(x,y)=(2,1)$.

**Verifica in entrambe le equazioni originali:**

- Prima: $2(2)+3(1)=4+3=7$. ✓
- Seconda: $3(2)-1=6-1=5$. ✓

**Perché moltiplichiamo per 3 (e non per un altro numero):** vogliamo che il coefficiente di $y$ nella seconda equazione ($-1$) diventi l'opposto esatto di quello nella prima ($3$). Moltiplicando $-1$ per $3$ otteniamo $-3$, l'opposto di $3$: è il numero minimo che rende i coefficienti opposti senza introdurre frazioni inutili.

````python
import sympy as sp

x, y = sp.symbols('x y')
sistema = [sp.Eq(2*x + 3*y, 7), sp.Eq(3*x - y, 5)]
soluzione = sp.solve(sistema, [x, y])
print("Soluzione:", soluzione)
````

## Riepilogo finale

- Un sistema di equazioni richiede di trovare valori che soddisfano **tutte** le equazioni contemporaneamente.
- **Sostituzione**: si ricava un'incognita da un'equazione e la si sostituisce nell'altra, riducendo a un'equazione in una sola incognita.
- **Eliminazione**: si moltiplicano le equazioni per opportuni numeri e le si sommano/sottraggono, in modo da far sparire un'incognita.
- Un sistema può essere impossibile (nessuna soluzione, equazioni "in contraddizione") o indeterminato (infinite soluzioni, equazioni equivalenti tra loro).
- Verifica sempre la soluzione trovata in **entrambe** le equazioni originali.
