# 1.11 — Esponenziali

**Prerequisiti:** 1.10 Potenze, 0.19 Relazioni e funzioni elementari.


## Perché ci interessa

Molti fenomeni naturali non crescono "a passi uguali", ma **si moltiplicano**: una popolazione di batteri che raddoppia ogni ora, un capitale che cresce di una percentuale fissa ogni anno (notebook 0.10), un isotopo radioattivo che dimezza la sua quantità in un tempo fisso. In tutti questi casi, la variabile compare **all'esponente**, non alla base.

Nel notebook precedente abbiamo esteso $a^x$ a ogni esponente reale. Ora studiamo $f(x)=a^x$ come vera **funzione** (notebook 0.19): dominio, grafico, comportamento — gli strumenti che ci serviranno per modellare crescita e decadimento in tutto il resto del corso.

## Teoria, passo dopo passo

### Definizione di funzione esponenziale

Fissata una base $a>0$, $a\ne1$, la **funzione esponenziale** di base $a$ è:

$$f(x) = a^x, \qquad x\in\mathbb{R}$$

Il dominio è **tutto $\mathbb{R}$** (grazie all'estensione vista nel notebook 1.10: $a^x$ ha senso per ogni $x$ reale, non solo razionale). Escludiamo $a=1$ perché $1^x=1$ per ogni $x$: sarebbe una funzione costante, poco interessante.

### Il grafico: due comportamenti possibili

Dalla monotonia vista nel notebook 1.10:

- Se $a>1$: il grafico **sale** da sinistra a destra (funzione crescente). Esempio tipico: $f(x)=2^x$.
- Se $0<a<1$: il grafico **scende** da sinistra a destra (funzione decrescente). Esempio tipico: $f(x)=\left(\frac12\right)^x$.

In entrambi i casi, alcune proprietà sono comuni:

- $f(0)=a^0=1$: ogni esponenziale passa per il punto $(0,1)$, qualunque sia la base.
- $f(x)>0$ per ogni $x$: una potenza con base positiva è sempre positiva (non è mai né zero né negativa). Il **codominio** (l'insieme dei valori assunti) è quindi $(0,+\infty)$, mai lo zero.
- L'asse $x$ (la retta $y=0$) è un **asintoto orizzontale**: il grafico si avvicina sempre di più all'asse $x$ senza mai toccarlo (per $x\to+\infty$ se $0<a<1$, oppure per $x\to-\infty$ se $a>1$). Il concetto preciso di asintoto lo formalizzeremo con i limiti, nel Modulo 4-5: per ora basti l'idea visiva.

### La base speciale $e$

Tra tutte le basi possibili, ce n'è una particolarmente importante in matematica: il numero $e\approx2.71828\dots$ (la lettera $e$ onora Eulero). È un numero irrazionale (come $\pi$, notebook 0.4), definito rigorosamente nel Modulo 4 tramite un limite di successione. La funzione $f(x)=e^x$, detta **esponenziale naturale**, ha proprietà speciali (legate alla derivata, che vedremo nel Modulo 5) che la rendono la base "naturale" per modellare crescita continua.

### Leggi di crescita e decadimento

Se $a>1$, diciamo che $f(x)=a^x$ modella una **crescita esponenziale**: raddoppiando $x$ di una quantità fissa, il valore si moltiplica sempre per lo stesso fattore (non si somma una quantità fissa, come nella crescita lineare del notebook 1.5). Se $0<a<1$, modella un **decadimento esponenziale**.

Esempio concreto: una popolazione di batteri che raddoppia ogni ora si modella con $N(t)=N_0\cdot2^t$, dove $N_0$ è la popolazione iniziale e $t$ il tempo in ore. Dopo 1 ora: $N_0\cdot2$. Dopo 2 ore: $N_0\cdot4$. Non è $N_0+\text{costante}$ (crescita lineare), ma $N_0\times\text{fattore}$ (crescita esponenziale) — una differenza concettuale fondamentale.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 200)

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(x, 2**x, label='f(x) = 2^x  (a>1, crescente)', color='tab:blue')
ax.plot(x, 0.5**x, label='f(x) = (1/2)^x  (0<a<1, decrescente)', color='tab:orange')
ax.plot(x, np.e**x, label='f(x) = e^x  (esponenziale naturale)', color='tab:green', linestyle='--')

ax.axhline(y=0, color='gray', linewidth=0.8)
ax.axvline(x=0, color='gray', linewidth=0.8)
ax.plot(0, 1, 'ko')  # tutte le esponenziali passano per (0,1)
ax.annotate('(0,1)', (0, 1), textcoords="offset points", xytext=(10, -15))

ax.set_ylim(-1, 10)
ax.legend()
ax.set_title("Funzioni esponenziali: tutte passano per (0,1), mai sotto l'asse x")
ax.grid(True, alpha=0.3)
plt.show()

# Verifica: tutte le esponenziali in x=0 valgono 1
for base in [2, 0.5, np.e, 10]:
    print(f"base={base:.4f}: f(0) = {base**0}")
````

## Esercizio svolto e commentato

**Problema:** un capitale di 1000 euro viene investito a un tasso di interesse composto del 5% annuo. Scrivi la funzione che descrive il capitale dopo $t$ anni, e calcola dopo quanti anni (circa) il capitale raddoppia.

**Ragionamento:**

**Impostiamo la funzione.** Un interesse del 5% annuo composto significa che ogni anno il capitale viene moltiplicato per $\left(1+\frac{5}{100}\right)=1.05$ (notebook 0.10, aumento percentuale). Dopo $t$ anni, il fattore si applica $t$ volte, cioè si eleva alla $t$ (esattamente come nell'esempio dei batteri, ma con base $1.05$ invece di $2$):

$$C(t) = 1000\times(1.05)^t$$

Questa è una funzione esponenziale con base $a=1.05>1$: crescita esponenziale.

**Troviamo quando raddoppia.** Vogliamo $C(t)=2000$, cioè:

$$1000\times(1.05)^t = 2000 \quad\Rightarrow\quad (1.05)^t=2$$

Qui ci troviamo di fronte a un problema nuovo: l'incognita $t$ è **all'esponente**. Non possiamo isolarla con gli strumenti visti finora (l'invariantiva, notebook 0.5, funziona per spostare termini, non per "tirare giù" un esponente). Ci serve uno strumento apposito — il **logaritmo** — che è esattamente l'argomento del prossimo notebook. Per ora, approssimiamo numericamente provando alcuni valori di $t$:

$$(1.05)^{10}\approx1.63 \qquad (1.05)^{14}\approx1.98 \qquad (1.05)^{15}\approx2.08$$

Il capitale raddoppia tra 14 e 15 anni: un valore preciso richiederà il logaritmo.

**Perché non possiamo "dividere per risolvere" come nelle equazioni lineari:** in $(1.05)^t=2$, $t$ non è moltiplicato per un numero (dove basterebbe dividere), ma è l'**esponente** di $1.05$. Serve un'operazione che "disfi" l'elevamento a potenza dalla parte dell'esponente — è precisamente il ruolo del logaritmo, che vedremo nel prossimo notebook.

````python
import numpy as np

def capitale(t, tasso=0.05, iniziale=1000):
    return iniziale * (1 + tasso)**t

for t in [10, 12, 14, 15, 16]:
    print(f"t={t} anni: capitale = {capitale(t):.2f} euro")

# Con SymPy possiamo comunque risolvere numericamente l'equazione
import sympy as sp
t = sp.symbols('t', positive=True)
soluzione = sp.nsolve(sp.Eq(1000*(1.05)**t, 2000), t, 14)
print(f"\nValore preciso di t (numerico): {soluzione}")
````

## Riepilogo finale

- La funzione esponenziale $f(x)=a^x$ ($a>0$, $a\ne1$) ha dominio tutto $\mathbb{R}$ e codominio $(0,+\infty)$: è sempre positiva.
- Se $a>1$ è crescente; se $0<a<1$ è decrescente; in entrambi i casi passa per $(0,1)$ e ha l'asse $x$ come asintoto orizzontale.
- Il numero $e\approx2.71828$ è la base "naturale" per la crescita continua, e la ritroveremo con un ruolo speciale nel Modulo 4 (limiti) e nel Modulo 5 (derivate).
- La crescita esponenziale (moltiplicare per un fattore fisso) è concettualmente diversa dalla crescita lineare (sommare una quantità fissa, notebook 1.5).
- Quando l'incognita compare all'esponente, l'invariantiva non basta più a isolarla: serve un nuovo strumento, il logaritmo.
