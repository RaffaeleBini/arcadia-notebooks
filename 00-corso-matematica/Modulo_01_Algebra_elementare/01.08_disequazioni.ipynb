# 1.8 — Disequazioni

**Prerequisiti:** 1.5 Equazioni di primo grado, 0.14 Intervalli, 0.15 Logica elementare.


## Perché ci interessa

Non sempre ci interessa "quando due quantità sono uguali", ma "quando una è maggiore (o minore) dell'altra": "per quali prezzi il profitto è positivo?", "per quali velocità il tempo di percorrenza è inferiore a un'ora?". Queste domande si traducono in **disequazioni**: relazioni con $<,\ >,\ \le,\ \ge$ al posto di $=$.

A differenza di un'equazione, la cui soluzione è tipicamente un numero (o pochi numeri), la soluzione di una disequazione è tipicamente un **intervallo** (notebook 0.14): un insieme continuo di valori.

## Teoria, passo dopo passo

### Le proprietà dell'ordine

Le disequazioni si risolvono con passaggi simili alle equazioni (invariantiva, notebook 0.5), ma con **una differenza cruciale**. Partiamo da una disuguaglianza vera, es. $3<5$, e vediamo cosa succede applicando operazioni:

- **Sommare/sottrarre uno stesso numero**: $3<5 \Rightarrow 3+2<5+2 \Rightarrow 5<7$. Vero: la disuguaglianza si conserva.
- **Moltiplicare/dividere per un numero positivo**: $3<5 \Rightarrow 3\times2<5\times2 \Rightarrow 6<10$. Vero: si conserva.
- **Moltiplicare/dividere per un numero negativo**: $3<5 \Rightarrow 3\times(-1) \ ? \ 5\times(-1) \Rightarrow -3 \ ? \ -5$. Ma $-3>-5$ (notebook 0.2, retta dei numeri): il verso della disuguaglianza **si è invertito**!

Questa è la regola fondamentale, spesso dimenticata: **moltiplicando o dividendo entrambi i membri di una disequazione per un numero negativo, il verso della disuguaglianza si inverte**. Il motivo è geometrico: moltiplicare per $-1$ è una riflessione della retta numerica rispetto allo zero (notebook 0.2), che scambia "sinistra" e "destra", cioè "minore" e "maggiore".

### Disequazioni di primo grado

Il procedimento è identico a quello delle equazioni di primo grado (notebook 1.5) — eliminare parentesi/frazioni, raggruppare, isolare $x$ — con l'unica accortezza di **invertire il verso** ogni volta che si moltiplica o divide per un numero negativo.

Esempio: risolvi $-2x+4>0$. Isoliamo $x$ sottraendo $4$: $-2x>-4$. Ora dividiamo per $-2$ (negativo!): il verso si inverte:

$$x<2$$

La soluzione è l'intervallo $(-\infty,2)$ (notebook 0.14).

### Come si scrive la soluzione

La soluzione di una disequazione in una incognita è quasi sempre un intervallo, o un'unione di intervalli. Si scrive usando esattamente la notazione del notebook 0.14: $x<2$ corrisponde a $(-\infty,2)$; $x\ge-1$ corrisponde a $[-1,+\infty)$.

### Sistemi di disequazioni

Quando più disequazioni devono valere **contemporaneamente** (proprio come "e" nella logica, notebook 0.15), la soluzione è l'**intersezione** dei singoli intervalli soluzione (esattamente come nel notebook 0.13, quando univamo condizioni con "e"):

$$\begin{cases} x>1 \\ x<5 \end{cases} \quad\Rightarrow\quad x\in(1,+\infty)\cap(-\infty,5) = (1,5)$$

Un modo pratico e visivo per trovare l'intersezione è disegnare ogni soluzione su una retta numerica (una sopra l'altra) e leggere la zona dove **tutte** sono soddisfatte contemporaneamente.

````python
import sympy as sp

x = sp.symbols('x')

# Verifica dell'inversione di verso moltiplicando per un negativo
print("3 < 5 ?", 3 < 5)
print("3*(-1) < 5*(-1) ?", 3*(-1) < 5*(-1))  # False: si è invertita
print("3*(-1) > 5*(-1) ?", 3*(-1) > 5*(-1))  # True: il verso corretto dopo l'inversione

# Risolviamo -2x+4>0 con SymPy
soluzione = sp.solve_univariate_inequality(-2*x + 4 > 0, x)
print("\nSoluzione di -2x+4>0:", soluzione)

# Sistema di disequazioni: intersezione degli intervalli
sol1 = sp.solve_univariate_inequality(x > 1, x)
sol2 = sp.solve_univariate_inequality(x < 5, x)
print(f"\nx>1: {sol1}")
print(f"x<5: {sol2}")
print(f"Intersezione (entrambe vere): {sol1.as_set().intersect(sol2.as_set())}")
````

## Esercizio svolto e commentato

**Problema:** risolvi il sistema $\begin{cases} 3x-6\le0 \\ -x+1<0 \end{cases}$.

**Ragionamento:**

Risolviamo ciascuna disequazione separatamente, poi intersechiamo le soluzioni.

**Prima disequazione — $3x-6\le0$:** sommiamo $6$ a entrambi i membri (invariantiva): $3x\le6$. Dividiamo per $3$ (positivo, il verso **non** cambia): $x\le2$. Soluzione: $(-\infty,2]$.

**Seconda disequazione — $-x+1<0$:** sottraiamo $1$: $-x<-1$. Dividiamo per $-1$ (negativo, il verso **si inverte**): $x>1$. Soluzione: $(1,+\infty)$.

**Intersezione delle due soluzioni:** vogliamo i valori che soddisfano **entrambe**, cioè $x\le2$ **e** $x>1$ contemporaneamente. Confrontando gli estremi (notebook 0.14): l'estremo sinistro più restrittivo è $1$ (escluso, dalla seconda), l'estremo destro più restrittivo è $2$ (incluso, dalla prima). Non c'è conflitto ($1<2$), quindi:

$$(-\infty,2] \cap (1,+\infty) = (1,2]$$

**Perché disegnare le due soluzioni su una retta aiuta a non sbagliare:** con più disequazioni è facile perdere il filo di quale estremo sia incluso o escluso. Segnando visivamente $(-\infty,2]$ e $(1,+\infty)$ su due rette numeriche allineate, la zona di sovrapposizione (l'intersezione) si legge a colpo d'occhio, riducendo il rischio di errore rispetto a un ragionamento solo "a mente".

````python
import sympy as sp

x = sp.symbols('x')
sol1 = sp.solve_univariate_inequality(3*x - 6 <= 0, x)
sol2 = sp.solve_univariate_inequality(-x + 1 < 0, x)
print("Prima disequazione:", sol1)
print("Seconda disequazione:", sol2)
print("Intersezione:", sol1.as_set().intersect(sol2.as_set()))
````

## Riepilogo finale

- Le disequazioni si risolvono come le equazioni (invariantiva, notebook 0.5), con una regola in più: **moltiplicando o dividendo per un numero negativo, il verso si inverte**.
- La soluzione è tipicamente un intervallo o un'unione di intervalli (notebook 0.14), non un singolo numero.
- Un sistema di disequazioni richiede che **tutte** valgano contemporaneamente: la soluzione è l'**intersezione** delle soluzioni singole.
- Disegnare le soluzioni su rette numeriche allineate aiuta a trovare l'intersezione senza errori, specialmente quando gli estremi inclusi/esclusi sono diversi.
