## Perché ci interessa

Abbiamo tutti gli strumenti pronti: nel notebook 0.8 sapevamo risolvere $|x-2|=5$ "sdoppiando" in due casi; nel notebook 1.14 abbiamo imparato a scrivere qualunque espressione con valore assoluto come funzione a tratti. Vedremo che esistono alcune "scorciatoie" per i casi più comuni ($|A|=k$, $|A|<k$, $|A|>k$), oltre al metodo generale degli intervalli, sempre disponibile quando le scorciatoie non bastano.

## Teoria, passo dopo passo

### Equazioni: $|A(x)|=k$

Se $k<0$: nessuna soluzione (un valore assoluto non è mai negativo, notebook 0.8). Se $k=0$: risolvi $A(x)=0$. Se $k>0$ (il caso generale, notebook 0.8):

$$|A(x)|=k \quad\Longleftrightarrow\quad A(x)=k \ \text{ oppure } \ A(x)=-k$$

### Equazioni: $|A(x)|=|B(x)|$

Due valori assoluti sono uguali se e solo se gli argomenti sono uguali **oppure opposti** (perché $|a|=|b|$ significa che $a$ e $b$ hanno la stessa distanza da zero, notebook 0.8 — e questo succede solo se sono lo stesso numero, o numeri opposti):

$$|A(x)|=|B(x)| \quad\Longleftrightarrow\quad A(x)=B(x) \ \text{ oppure } \ A(x)=-B(x)$$

Questa è una scorciatoia comoda: **non serve** lo studio del segno per questo caso specifico, a differenza del caso generale.

### Disequazioni: $|A(x)|<k$ e $|A(x)|>k$ (con $k>0$)

Ricordiamo l'intuizione del notebook 0.8: $|A(x)|<k$ significa "la distanza di $A(x)$ da zero è minore di $k$", cioè $A(x)$ è compreso strettamente tra $-k$ e $k$:

$$|A(x)|<k \quad\Longleftrightarrow\quad -k<A(x)<k$$

Si trasforma quindi in un **sistema** di due disequazioni (notebook 1.8), da intersecare. Per $|A(x)|>k$, vale il contrario: $A(x)$ deve stare **fuori** dall'intervallo $[-k,k]$:

$$|A(x)|>k \quad\Longleftrightarrow\quad A(x)<-k \ \text{ oppure } \ A(x)>k$$

Qui la soluzione è l'**unione**, non l'intersezione (notebook 0.13): basta che valga una delle due condizioni.

**Perché "minore" dà un'intersezione e "maggiore" dà un'unione — un punto spesso confuso:** pensa alla retta numerica. Stare "vicino a zero" (distanza piccola) è una condizione che stringe $A(x)$ in una zona centrale limitata (da cui il sistema/intersezione). Stare "lontano da zero" (distanza grande) permette ad $A(x)$ di scappare verso $+\infty$ **oppure** verso $-\infty$: due possibilità alternative (da cui l'unione).

### Il metodo generale: studio del segno (quando le scorciatoie non bastano)

Per casi più complessi — più valori assoluti sommati, o disequazioni non riconducibili alle forme sopra — si applica il metodo generale del notebook 1.14: si individuano i punti critici, si riscrive l'espressione senza valori assoluti in ciascun intervallo, e si risolve la disequazione (algebrica, senza più valore assoluto) **separatamente in ciascun intervallo**, ricordando di intersecare ogni soluzione parziale con l'intervallo in cui è stata trovata. La soluzione finale è l'unione di tutte le soluzioni parziali.

````python
import sympy as sp

x = sp.symbols('x', real=True)

# |x-3| = 5
sol1 = sp.solve(sp.Eq(sp.Abs(x - 3), 5), x)
print("|x-3|=5:", sol1)

# |2x-1| = |x+4|
sol2 = sp.solve(sp.Eq(sp.Abs(2*x - 1), sp.Abs(x + 4)), x)
print("|2x-1|=|x+4|:", sol2)

# |x-1| < 3  (deve dare un intervallo aperto)
sol3 = sp.solve_univariate_inequality(sp.Abs(x - 1) < 3, x)
print("|x-1|<3:", sol3)

# |x+2| > 4  (deve dare un'unione di due semirette)
sol4 = sp.solve_univariate_inequality(sp.Abs(x + 2) > 4, x)
print("|x+2|>4:", sol4)
````

## Esercizio svolto e commentato

**Problema:** risolvi $|x-1|+|x+2|\le5$ (usando l'espressione già scomposta a tratti nel notebook 1.14).

**Ragionamento:**

Riprendiamo dal notebook 1.14 la scomposizione a tratti già trovata per $|x-1|+|x+2|$:

$$|x-1|+|x+2| = \begin{cases} -2x-1 & x<-2 \\ 3 & -2\le x<1 \\ 2x+1 & x\ge1 \end{cases}$$

Risolviamo la disequazione $\le5$ **separatamente in ciascun intervallo**, poi uniamo i risultati.

**Intervallo $x<-2$:** risolviamo $-2x-1\le5$ (equazione algebrica ordinaria, notebook 1.8): $-2x\le6 \Rightarrow x\ge-3$ (verso invertito, dividiamo per un negativo). Intersechiamo con l'intervallo di partenza $x<-2$: $-3\le x<-2$.

**Intervallo $-2\le x<1$:** l'espressione vale costantemente $3$, e $3\le5$ è **sempre vera**. Quindi tutto l'intervallo è soluzione: $-2\le x<1$.

**Intervallo $x\ge1$:** risolviamo $2x+1\le5$: $2x\le4 \Rightarrow x\le2$. Intersechiamo con $x\ge1$: $1\le x\le2$.

**Uniamo le tre soluzioni parziali** (notebook 0.13):

$$[-3,-2) \cup [-2,1) \cup [1,2] = [-3,2]$$

(i tre pezzi sono "adiacenti senza buchi", quindi si fondono in un unico intervallo chiuso).

**Verifica dell'interpretazione geometrica**: $|x-1|+|x+2|$ è la somma delle distanze da $1$ e da $-2$ (notebook 1.14). Il valore $5$ è raggiunto esattamente agli estremi $x=-3$ e $x=2$ (a distanza $3$ in più rispetto ai punti critici, ripartita simmetricamente): coerente con il risultato $[-3,2]$.

**Perché risolvere "un pezzo alla volta, poi unire" è l'unico modo affidabile quando l'espressione cambia formula:** non ha senso applicare un'unica formula globale, perché letteralmente non esiste — l'espressione *è* diversa in ogni intervallo. Il metodo a tratti, per quanto richieda più passaggi, è l'unico che rispetta fedelmente la vera natura dell'espressione.

````python
import sympy as sp

x = sp.symbols('x', real=True)
soluzione = sp.solve_univariate_inequality(sp.Abs(x - 1) + sp.Abs(x + 2) <= 5, x)
print("Soluzione:", soluzione)
````

## Riepilogo finale

- $|A(x)|=k$ (con $k>0$) si sdoppia in $A(x)=k$ oppure $A(x)=-k$; $|A(x)|=|B(x)|$ si sdoppia in $A(x)=B(x)$ oppure $A(x)=-B(x)$, senza bisogno dello studio del segno.
- $|A(x)|<k$ diventa il sistema $-k<A(x)<k$ (**intersezione**); $|A(x)|>k$ diventa $A(x)<-k$ oppure $A(x)>k$ (**unione**) — ricorda: "vicino" stringe, "lontano" apre in due direzioni.
- Per espressioni più complesse, il metodo generale è quello a tratti (notebook 1.14): risolvere separatamente in ogni intervallo definito dai punti critici, poi unire tutte le soluzioni parziali.
