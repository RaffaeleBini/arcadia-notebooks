# 1.10 — Potenze

**Prerequisiti:** 0.6 Frazioni, potenze e radici.


## Perché ci interessa

Nel notebook 0.6 abbiamo definito $a^n$ per $n$ naturale, poi esteso a $n$ intero negativo e a $n$ razionale (tramite le radici, $a^{1/n}=\sqrt[n]{a}$). Ma cosa significa $2^{\sqrt{2}}$? L'esponente $\sqrt{2}$ è un numero **irrazionale** (notebook 0.4): non possiamo interpretarlo come "moltiplicare $2$ per sé stesso un certo numero di volte", né come una radice con indice intero.

Estendere le potenze a esponenti reali qualsiasi è il passo che ci serve per costruire, nel prossimo notebook, le **funzioni esponenziali** — funzioni definite per ogni $x$ reale, non solo per gli $x$ razionali.

## Teoria, passo dopo passo

### Cosa sappiamo già

Ricapitoliamo dal notebook 0.6: per $a>0$ e $p,q$ interi con $q\ne0$,

$$a^{p/q} = \sqrt[q]{a^p}$$

Questo ci dà significato a $a^x$ per ogni $x$ **razionale**. I razionali sono densi in $\mathbb{R}$ (notebook 0.3): tra due reali distinti c'è sempre un razionale. Questo suggerisce un'idea per definire $a^x$ anche quando $x$ è irrazionale: **approssimare**.

### L'idea di approssimazione

Consideriamo $2^{\sqrt{2}}$. Sappiamo che $\sqrt{2}=1.41421356\dots$ Possiamo calcolare le potenze con esponenti razionali sempre più vicini a $\sqrt{2}$:

$$2^{1.4},\quad 2^{1.41},\quad 2^{1.414},\quad 2^{1.4142},\ \dots$$

Ognuno di questi è un numero che sappiamo calcolare esattamente (esponente razionale, notebook 0.6). Man mano che l'esponente si avvicina a $\sqrt{2}$, questi valori si avvicinano sempre di più a un unico numero reale, che **definiamo** come $2^{\sqrt{2}}$. Questa idea — costruire un valore come "limite" di approssimazioni sempre più precise — è un'anticipazione informale del concetto di **limite**, che formalizzeremo con rigore nel Modulo 4 e nel Modulo 5. Per ora ci basta l'intuizione: la successione di valori "si stabilizza" su un numero preciso, e quello è $2^{\sqrt{2}}$.

### Le proprietà restano le stesse

Il punto cruciale è che questa estensione **mantiene valide** tutte le proprietà delle potenze viste nel notebook 0.6, ora per esponenti reali qualsiasi $x,y\in\mathbb{R}$ (con $a,b>0$):

$$a^x\cdot a^y=a^{x+y} \qquad \frac{a^x}{a^y}=a^{x-y} \qquad (a^x)^y=a^{xy} \qquad (ab)^x=a^xb^x$$

Non è un caso: la costruzione per approssimazione è fatta apposta per preservarle (le proprietà valgono su ogni approssimazione razionale, quindi continuano a valere anche nel limite).

### Perché serve $a>0$

Se $a<0$, per esempio $a=-1$, l'espressione $a^x$ diventa problematica per esponenti non interi: $(-1)^{1/2}=\sqrt{-1}$ non è un numero reale (notebook 0.4). Per questo, quando parliamo di potenze con esponente reale qualsiasi, richiediamo sempre **base positiva**: $a>0$. (Il caso $a=0$ si tratta a parte: $0^x=0$ per $x>0$, mentre $0^0$ è una forma che si lascia indefinita o si definisce per convenzione a seconda del contesto — non ce ne occuperemo qui.)

### Il comportamento qualitativo di $a^x$

Fissata una base $a>0$, $a\ne1$, osserviamo come cambia $a^x$ al crescere di $x$:

- Se $a>1$: $a^x$ **cresce** al crescere di $x$ (es. $2^1=2$, $2^2=4$, $2^3=8$, \dots).
- Se $0<a<1$: $a^x$ **decresce** al crescere di $x$ (es. $\left(\frac12\right)^1=\frac12$, $\left(\frac12\right)^2=\frac14$, \dots).
- Se $a=1$: $a^x=1$ per ogni $x$ (caso banale, non interessante).

Questa osservazione qualitativa sarà il punto di partenza del prossimo notebook, dove studieremo $f(x)=a^x$ come vera e propria funzione.

````python
import sympy as sp
import numpy as np

# Approssimiamo 2^sqrt(2) con esponenti razionali sempre più precisi
approssimazioni_esponente = [1.4, 1.41, 1.414, 1.4142, 1.41421356]
print("Approssimazioni di 2^sqrt(2):")
for e in approssimazioni_esponente:
    print(f"  2^{e} = {2**e:.8f}")

valore_esatto = 2 ** np.sqrt(2)
print(f"\n2^sqrt(2) (calcolato direttamente) = {valore_esatto:.8f}")

# Verifica simbolica che le proprietà delle potenze restano valide con SymPy
a, x, y = sp.symbols('a x y', positive=True)
proprieta1 = sp.simplify(a**x * a**y - a**(x + y))
print("\na^x * a^y - a^(x+y) =", proprieta1, "(deve fare 0)")

# Confronto qualitativo: a>1 cresce, 0<a<1 decresce
valori_x = [0, 1, 2, 3]
print("\nCon a=2 (>1), a^x cresce:", [2**v for v in valori_x])
print("Con a=0.5 (<1), a^x decresce:", [0.5**v for v in valori_x])
````

## Esercizio svolto e commentato

**Problema:** senza calcolare il valore esatto, stabilisci se $3^{\pi}$ è maggiore o minore di $3^{3}$ e di $3^{3.5}$, motivando con le proprietà viste.

**Ragionamento:**

Sappiamo che $\pi=3.14159\dots$, quindi:

$$3 < \pi < 3.5$$

La base è $a=3$, e $3>1$: nella teoria abbiamo osservato che, per basi maggiori di 1, $a^x$ **cresce** al crescere di $x$. Questo significa che a esponenti più grandi corrispondono valori più grandi (la funzione "rispetta l'ordine").

Applicando questa proprietà di monotonia ai tre esponenti $3<\pi<3.5$:

$$3^3 < 3^{\pi} < 3^{3.5}$$

Quindi $3^{\pi}$ è compreso strettamente tra $3^3=27$ e $3^{3.5}$, senza bisogno di calcolarne il valore esatto.

**Perché possiamo confrontare gli esponenti invece di calcolare i valori:** questo è esattamente il senso di dire che $a^x$ è **crescente** per $a>1$: se sappiamo ordinare gli esponenti, sappiamo automaticamente ordinare le potenze corrispondenti, senza fare nessun calcolo numerico. Questa proprietà di monotonia sarà centrale nel prossimo notebook, quando studieremo le funzioni esponenziali e le loro proprietà.

````python
import numpy as np

valori = {
    "3^3": 3**3,
    "3^pi": 3**np.pi,
    "3^3.5": 3**3.5,
}
for nome, valore in valori.items():
    print(f"{nome} = {valore:.4f}")

print("\nOrdine crescente confermato:", valori["3^3"] < valori["3^pi"] < valori["3^3.5"])
````

## Riepilogo finale

- $a^x$ per $x$ irrazionale si definisce approssimando $x$ con esponenti razionali sempre più precisi: un'anticipazione informale del concetto di limite (Moduli 4-5).
- Le proprietà delle potenze (notebook 0.6) restano valide per esponenti reali qualsiasi, a patto che la base sia $a>0$.
- Se $a>1$, $a^x$ è **crescente**; se $0<a<1$, $a^x$ è **decrescente**: questa monotonia permette di confrontare potenze senza calcolarle esplicitamente.
- Questa estensione delle potenze a tutto $\mathbb{R}$ è ciò che rende possibile definire $f(x)=a^x$ come funzione su tutta la retta reale, non solo sui razionali.
