**Prerequisiti:** 1.3 Scomposizione dei polinomi, 0.3 Numeri razionali.


## Perché ci interessa

Nel notebook 0.3 abbiamo definito i numeri razionali come rapporti tra interi, $\frac{a}{b}$ con $b\ne0$. Una **frazione algebrica** è la stessa idea, ma con polinomi al posto di numeri: $\frac{x+1}{x^2-4}$.

C'è però una differenza cruciale rispetto alle frazioni numeriche: qui il denominatore contiene una lettera, quindi **potrebbe annullarsi** per certi valori di $x$. Dobbiamo imparare a riconoscere questi valori "proibiti" prima di poter lavorare con la frazione — un'attenzione che tornerà utile per tutto il resto del corso, ogni volta che vedremo una lettera al denominatore.

## Teoria, passo dopo passo

### Il dominio di una frazione algebrica

Una frazione algebrica $\frac{P(x)}{Q(x)}$ (con $P,Q$ polinomi) è definita per tutti i valori di $x$ **tranne** quelli che annullano il denominatore $Q(x)$. L'insieme dei valori ammessi si chiama **dominio** (o **condizioni di esistenza**, C.E.).

Esempio: $\frac{x+1}{x-3}$ è definita per ogni $x$ tranne $x=3$ (perché $3-3=0$, e la divisione per zero non è mai ammessa, notebook 0.3). Scriviamo: C.E. $x\ne3$.

Se il denominatore è un polinomio di grado più alto, bisogna prima scomporlo (notebook 1.3) per trovare **tutti** i valori che lo annullano:

$$\frac{x+1}{x^2-4} = \frac{x+1}{(x+2)(x-2)} \qquad \text{C.E. } x\ne-2,\ x\ne2$$

**Perché scomporre il denominatore è indispensabile:** guardando $x^2-4$ non scomposto, non è immediato vedere per quali $x$ si annulla. Scomposto come $(x+2)(x-2)$, ricordiamo che un prodotto è zero se e solo se almeno uno dei fattori è zero (questa proprietà, che useremo moltissimo, si chiama **legge di annullamento del prodotto** e la riprenderemo nel notebook 1.6): quindi il denominatore si annulla quando $x+2=0$ oppure $x-2=0$, cioè $x=-2$ oppure $x=2$.

### Semplificare una frazione algebrica

Come con le frazioni numeriche (notebook 0.3), si semplifica dividendo numeratore e denominatore per un fattore comune — ma qui il fattore comune si trova **scomponendo** entrambi:

$$\frac{x^2-4}{x^2-2x} = \frac{(x+2)(x-2)}{x(x-2)} = \frac{x+2}{x}$$

**Attenzione**: la semplificazione è valida solo per i valori di $x$ nel dominio originale. Qui il dominio richiede $x\ne0$ e $x\ne2$ (dal denominatore originale $x^2-2x=x(x-2)$): anche dopo aver semplificato il fattore $(x-2)$, dobbiamo **ricordare** che $x=2$ resta escluso, anche se la frazione semplificata $\frac{x+2}{x}$ non lo "segnala" più esplicitamente.

### Moltiplicazione e divisione di frazioni algebriche

Esattamente come per le frazioni numeriche (notebook 0.3): si moltiplicano numeratori tra loro e denominatori tra loro, poi si semplifica scomponendo:

$$\frac{x+1}{x-2} \times \frac{x-2}{x+3} = \frac{(x+1)(x-2)}{(x-2)(x+3)} = \frac{x+1}{x+3} \qquad (x\ne2)$$

La divisione si trasforma in moltiplicazione per il reciproco, come nel notebook 0.3.

### Addizione e sottrazione di frazioni algebriche

Come per le frazioni numeriche, serve un denominatore comune. Il modo più efficiente è scomporre ogni denominatore e trovare il minimo comune multiplo tra i fattori:

$$\frac{1}{x} + \frac{1}{x+1} = \frac{(x+1) + x}{x(x+1)} = \frac{2x+1}{x(x+1)} \qquad (x\ne0,\ x\ne-1)$$

````python
import sympy as sp

x = sp.symbols('x')

# Dominio: SymPy può trovare dove si annulla il denominatore
denominatore = x**2 - 4
zeri = sp.solve(sp.Eq(denominatore, 0), x)
print("Il denominatore x^2-4 si annulla per x =", zeri)

# Semplificazione di una frazione algebrica
frazione = (x**2 - 4) / (x**2 - 2*x)
semplificata = sp.simplify(frazione)
print("\n(x^2-4)/(x^2-2x) semplificata:", semplificata)

# Moltiplicazione di frazioni algebriche
prodotto = sp.simplify(((x + 1) / (x - 2)) * ((x - 2) / (x + 3)))
print("\nProdotto semplificato:", prodotto)

# Addizione di frazioni algebriche
somma = sp.together(1/x + 1/(x + 1))  # together porta a denominatore comune
print("\n1/x + 1/(x+1) =", somma)
````

## Esercizio svolto e commentato

**Problema:** semplifica $\dfrac{x^2-x-6}{x^2-9}$, indicando le condizioni di esistenza.

**Ragionamento:**

**Passo 1 — scomponiamo il numeratore.** $x^2-x-6$ è un trinomio (notebook 1.3): cerchiamo due numeri che sommati diano $-1$ e moltiplicati diano $-6$. Sono $-3$ e $2$. Quindi:

$$x^2-x-6 = (x-3)(x+2)$$

**Passo 2 — scomponiamo il denominatore.** $x^2-9$ è una differenza di quadrati (notebook 1.2):

$$x^2-9 = (x-3)(x+3)$$

**Passo 3 — troviamo le condizioni di esistenza**, cioè i valori che annullano il denominatore *originale* (prima di semplificare): $x-3=0$ oppure $x+3=0$, quindi C.E. $x\ne3,\ x\ne-3$.

**Passo 4 — riscriviamo la frazione scomposta e semplifichiamo** il fattore comune $(x-3)$:

$$\frac{(x-3)(x+2)}{(x-3)(x+3)} = \frac{x+2}{x+3} \qquad (x\ne3,\ x\ne-3)$$

**Perché scriviamo ancora "$x\ne3$" anche se nella frazione semplificata $x=3$ non causerebbe più problemi apparenti:** la frazione semplificata $\frac{x+2}{x+3}$ e quella originale $\frac{x^2-x-6}{x^2-9}$ **non sono la stessa funzione**: in $x=3$, l'originale non è definita (denominatore zero), mentre la semplificata varrebbe $\frac{5}{6}$. Le condizioni di esistenza vanno sempre riportate insieme al risultato, altrimenti si perde un'informazione essenziale sul dominio.

````python
import sympy as sp

x = sp.symbols('x')
frazione = (x**2 - x - 6) / (x**2 - 9)
semplificata = sp.simplify(frazione)
print("Frazione semplificata:", semplificata)

# Condizioni di esistenza (dal denominatore originale)
ce = sp.solve(sp.Eq(x**2 - 9, 0), x)
print("Valori esclusi (C.E.):", ce)
````

## Riepilogo finale

- Una frazione algebrica $\frac{P(x)}{Q(x)}$ è definita per ogni $x$ che non annulla $Q(x)$: le **condizioni di esistenza** (C.E.) vanno sempre determinate scomponendo il denominatore.
- Per semplificare, si scompongono numeratore e denominatore (notebook 1.3) e si eliminano i fattori comuni — ma le C.E. calcolate sul denominatore **originale** restano valide e vanno sempre riportate.
- Moltiplicazione, divisione, addizione e sottrazione seguono le stesse regole delle frazioni numeriche (notebook 0.3), ma con la scomposizione al posto della semplice ricerca del MCD.
- Una frazione algebrica semplificata e quella originale non sono la stessa funzione: differiscono nei punti esclusi dal dominio.
