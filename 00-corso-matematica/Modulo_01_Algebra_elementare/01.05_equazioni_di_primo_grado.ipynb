# 1.5 — Equazioni di primo grado

**Prerequisiti:** 0.5 Proprietà delle operazioni (in particolare l'invariantiva), 1.1 Polinomi.


## Perché ci interessa

Nel notebook 0.5 abbiamo già risolto $2(x+3)=14$ passo per passo, usando esplicitamente la proprietà invariantiva. Quell'esempio **era già** un'equazione di primo grado: qui sistematizziamo il procedimento, gli diamo un nome, e lo applichiamo a casi via via più generali.

Le equazioni di primo grado (dette anche **lineari**) sono lo strumento algebrico più usato in assoluto: ogni volta che una relazione tra grandezze è "proporzionale" o "diretta" (senza quadrati, radici o altre complicazioni), risolverla significa risolvere un'equazione di questo tipo.

## Teoria, passo dopo passo

### Cos'è un'equazione di primo grado

Un'**equazione** è un'uguaglianza tra due espressioni contenenti un'incognita (di solito $x$), vera solo per certi valori di $x$ (le **soluzioni**, o **radici**). Un'equazione di primo grado in una incognita si può sempre ricondurre, con passaggi leciti, alla forma:

$$ax+b=0 \qquad (a\ne0)$$

dove $a,b$ sono numeri (coefficienti). Il nome "primo grado" viene dal fatto che l'incognita compare solo con esponente 1 (nessun $x^2$, nessuna radice di $x$).

### Risolvere con la proprietà invariantiva

Partendo da $ax+b=0$, isoliamo $x$ con due applicazioni dell'invariantiva (notebook 0.5):

$$ax+b=0 \ \xrightarrow{-b \text{ a entrambi i membri}}\ ax=-b \ \xrightarrow{\div a \text{ a entrambi i membri}}\ x=-\frac{b}{a}$$

La divisione per $a$ è lecita perché abbiamo richiesto $a\ne0$: è lo stesso motivo per cui, nella proprietà invariantiva della moltiplicazione (notebook 0.5), il numero per cui si moltiplica/divide deve essere diverso da zero.

### La forma generale e i passaggi tipici

Un'equazione di primo grado si presenta di solito in una forma più complicata, con l'incognita sparsa su entrambi i lati e magari dentro delle parentesi:

$$3(x-2) = 5x+4$$

La strategia è sempre la stessa, in tre fasi:

1. **Eliminare le parentesi** con la distributiva (notebook 1.1): $3x-6=5x+4$.
2. **Raggruppare i termini con $x$ da un lato e i numeri dall'altro**, usando l'invariantiva (sottraendo $5x$ e sommando $6$ a entrambi i membri): $3x-5x=4+6$, cioè $-2x=10$.
3. **Isolare $x$** dividendo per il coefficiente: $x=\dfrac{10}{-2}=-5$.

**Verifica sempre la soluzione**, sostituendola nell'equazione originale: $3(-5-2) = 3(-7)=-21$, e $5(-5)+4=-25+4=-21$. Coincidono. ✓

### Casi particolari: nessuna soluzione o infinite soluzioni

Cosa succede se, dopo aver raggruppato, il coefficiente di $x$ diventa $0$? Consideriamo $2x+3=2x+7$. Raggruppando: $2x-2x=7-3$, cioè $0=4$. Questa è un'affermazione **falsa** (per qualunque $x$): l'equazione non ha soluzione, si dice **impossibile**.

Consideriamo invece $2x+3=2x+3$: raggruppando otteniamo $0=0$, un'affermazione **sempre vera**: l'equazione è verificata da **ogni** valore di $x$, si dice **indeterminata** (o identità).

Questi due casi speciali si riconoscono proprio quando, isolando $x$, il coefficiente davanti a $x$ si annulla — analogo a quanto visto nel notebook 1.1 sull'esercizio del grado che "crolla" quando un coefficiente si annulla.

````python
import sympy as sp

x = sp.symbols('x')

# Risolviamo 3(x-2) = 5x+4
equazione = sp.Eq(3*(x - 2), 5*x + 4)
soluzione = sp.solve(equazione, x)
print("Soluzione di 3(x-2)=5x+4:", soluzione)

# Verifica sostituendo il valore trovato
valore = soluzione[0]
lhs = 3*(valore - 2)
rhs = 5*valore + 4
print(f"Verifica: membro sinistro = {lhs}, membro destro = {rhs}")

# Caso impossibile: 2x+3 = 2x+7
equazione_impossibile = sp.Eq(2*x + 3, 2*x + 7)
print("\nSoluzioni di 2x+3=2x+7:", sp.solve(equazione_impossibile, x))  # lista vuota

# Caso indeterminato: 2x+3 = 2x+3
equazione_identita = sp.Eq(2*x + 3, 2*x + 3)
print("2x+3=2x+3 è sempre vera?", equazione_identita)
print("simplify:", sp.simplify(equazione_identita))  # True: vera per ogni x
````

## Esercizio svolto e commentato

**Problema:** risolvi $\dfrac{x-1}{2} + \dfrac{x+1}{3} = 2$, mostrando ogni passaggio.

**Ragionamento:**

Questa equazione contiene frazioni: la strategia più pulita è **eliminarle subito**, moltiplicando entrambi i membri per il denominatore comune (invariantiva della moltiplicazione, notebook 0.5).

**Passo 1 — troviamo il denominatore comune** tra $2$ e $3$: è $6$ (notebook 0.3).

**Passo 2 — moltiplichiamo entrambi i membri per 6:**

$$6\times\left(\frac{x-1}{2} + \frac{x+1}{3}\right) = 6\times2$$

Distribuendo il $6$ (proprietà distributiva, notebook 0.5) su ciascun termine a sinistra:

$$6\times\frac{x-1}{2} + 6\times\frac{x+1}{3} = 12$$

$$3(x-1) + 2(x+1) = 12$$

Ora l'equazione non ha più frazioni.

**Passo 3 — eliminiamo le parentesi** con la distributiva:

$$3x-3+2x+2=12$$

**Passo 4 — raggruppiamo** i termini con $x$ e i numeri:

$$3x+2x = 12+3-2 \quad \Rightarrow \quad 5x=13$$

**Passo 5 — isoliamo $x$:**

$$x=\frac{13}{5}$$

**Perché moltiplicare per il denominatore comune *prima* di ogni altro passaggio è la scelta migliore:** lavorare con le frazioni fino alla fine (sommandole man mano) è possibile ma più macchinoso e più a rischio di errore. Eliminarle subito, moltiplicando tutta l'equazione per il denominatore comune, trasforma il problema in un'equazione con soli numeri interi, molto più semplice da maneggiare.

````python
import sympy as sp

x = sp.symbols('x')
equazione = sp.Eq((x - 1)/2 + (x + 1)/3, 2)
soluzione = sp.solve(equazione, x)
print("Soluzione:", soluzione)  # atteso 13/5
````

## Riepilogo finale

- Un'equazione di primo grado si riconduce sempre alla forma $ax+b=0$ (con $a\ne0$), risolta isolando $x$ con l'invariantiva (notebook 0.5): $x=-\frac{b}{a}$.
- Il procedimento standard: eliminare parentesi e frazioni (moltiplicando per il denominatore comune), raggruppare i termini con $x$ da un lato e i numeri dall'altro, isolare $x$.
- Se il coefficiente di $x$ si annulla dopo il raggruppamento, l'equazione è **impossibile** (nessuna soluzione, es. $0=4$) oppure **indeterminata** (infinite soluzioni, es. $0=0$).
- Verificare sempre la soluzione trovata sostituendola nell'equazione originale.
