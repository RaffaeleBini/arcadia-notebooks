In questo notebook sistematizziamo il metodo generale — lo **studio del segno degli argomenti**, che generalizza quanto già visto nel notebook 1.9 — per gestire espressioni con valore assoluto di qualunque complessità, in vista delle equazioni e disequazioni del prossimo notebook.

## Teoria, passo dopo passo

### Il valore assoluto come funzione definita "a tratti"

Ricordiamo la definizione (notebook 0.8): $|x|=x$ se $x\ge0$, $|x|=-x$ se $x<0$. Possiamo applicare la stessa idea a un'espressione qualsiasi $f(x)$ al posto di $x$:

$$|f(x)| = \begin{cases} f(x) & \text{se } f(x)\ge0 \\ -f(x) & \text{se } f(x)<0 \end{cases}$$

Questo si chiama definire una funzione **a tratti** (o "per casi"): la formula cambia a seconda di dove ci troviamo. Per applicarla, dobbiamo prima capire **dove** $f(x)$ è positivo o negativo — esattamente lo studio del segno del notebook 1.9!

Esempio: $|2x-1|$. Studiamo il segno di $2x-1$: è $\ge0$ quando $x\ge\frac12$, è $<0$ quando $x<\frac12$. Quindi:

$$|2x-1| = \begin{cases} 2x-1 & \text{se } x\ge\frac12 \\ -(2x-1)=1-2x & \text{se } x<\frac12 \end{cases}$$

### Il grafico di $|f(x)|$ a partire dal grafico di $f(x)$

C'è un modo visivo rapido per ottenere il grafico di $|f(x)|$ conoscendo quello di $f(x)$: **dove $f(x)\ge0$, il grafico di $|f(x)|$ coincide con quello di $f(x)$; dove $f(x)<0$, il grafico di $|f(x)|$ è il "ribaltamento" (riflessione rispetto all'asse $x$)** del grafico di $f(x)$ in quella zona. Questo segue direttamente dalla definizione: dove $f(x)<0$, $|f(x)|=-f(x)$, che è precisamente il ribaltato di $f(x)$.

Visivamente: il grafico di $|f(x)|$ sta sempre "sopra o sull'asse x" (mai sotto, perché un valore assoluto non è mai negativo, notebook 0.8), ed è ottenuto "piegando verso l'alto" tutte le parti del grafico di $f(x)$ che stavano sotto l'asse.

### Più valori assoluti nella stessa espressione: il metodo degli intervalli

Quando un'espressione contiene **più** valori assoluti, ognuno "cambia comportamento" nel proprio punto critico (dove il suo argomento si annulla). Il metodo generale:

1. Trova i punti critici di **ciascun** valore assoluto (dove ogni argomento si annulla).
2. Ordina questi punti sulla retta numerica: dividono $\mathbb{R}$ in intervalli.
3. In **ciascun** intervallo, ogni argomento ha un segno costante (positivo o negativo): riscrivi l'espressione senza valori assoluti, sostituendo ciascuno con $+f(x)$ o $-f(x)$ a seconda del segno in quell'intervallo.
4. Risolvi (o semplifica) separatamente in ogni intervallo.

Esempio: $|x-1|+|x+2|$. Punti critici: $x=1$ (da $x-1$) e $x=-2$ (da $x+2$). Questi dividono la retta in tre intervalli: $x<-2$, $-2\le x<1$, $x\ge1$.

| Intervallo | segno di $x-1$ | segno di $x+2$ | $|x-1|+|x+2|$ |
|---|---|---|---|
| $x<-2$ | $-$ | $-$ | $(1-x)+(-x-2) = -2x-1$ |
| $-2\le x<1$ | $-$ | $+$ | $(1-x)+(x+2) = 3$ |
| $x\ge1$ | $+$ | $+$ | $(x-1)+(x+2) = 2x+1$ |

Notiamo che nell'intervallo centrale l'espressione **diventa costante** ($=3$): questo è il valore minimo, e corrisponde esattamente alla distanza tra i due punti critici $-2$ e $1$ (interpretazione geometrica: $|x-1|+|x+2|$ è la somma delle distanze di $x$ da $1$ e da $-2$, minima quando $x$ sta *tra* i due punti, notebook 0.8).

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

# Verifica della scrittura a tratti di |2x-1|
espressione = sp.Abs(2*x - 1)
print("|2x-1| scritto a tratti da SymPy:")
sp.pprint(sp.piecewise_fold(espressione.rewrite(sp.Piecewise)))

# Grafico di f(x) = x^2-1 e di |f(x)|, per vedere il "ribaltamento"
x_vals = np.linspace(-3, 3, 300)
f = x_vals**2 - 1
abs_f = np.abs(f)

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(x_vals, f, label='f(x) = x²-1', color='tab:blue')
ax.plot(x_vals, abs_f, label='|f(x)| = |x²-1|', color='tab:red', linestyle='--')
ax.axhline(y=0, color='gray', linewidth=0.8)
ax.legend()
ax.set_title("La parte di f(x) sotto l'asse x viene 'ribaltata' in |f(x)|")
ax.grid(True, alpha=0.3)
plt.show()

# Verifica dell'esempio |x-1|+|x+2|: minimo costante tra i punti critici
def somma_distanze(x_val):
    return abs(x_val - 1) + abs(x_val + 2)

for x_val in [-5, -2, -1, 0, 1, 4]:
    print(f"x={x_val}: |x-1|+|x+2| = {somma_distanze(x_val)}")
````

## Esercizio svolto e commentato

**Problema:** scrivi $|x+3| - |2x-4|$ come funzione a tratti (senza valori assoluti), individuando ogni intervallo.

**Ragionamento:**

**Passo 1 — troviamo i punti critici.** Il primo valore assoluto si annulla quando $x+3=0$, cioè $x=-3$. Il secondo quando $2x-4=0$, cioè $x=2$.

**Passo 2 — ordiniamo i punti critici e definiamo gli intervalli**: $-3<2$, quindi gli intervalli sono $x<-3$, $-3\le x<2$, $x\ge2$.

**Passo 3 — determiniamo il segno di ciascun argomento in ogni intervallo:**

- Per $x<-3$: $x+3<0$ (quindi $|x+3|=-(x+3)$), e $2x-4<0$ ancora più chiaramente, essendo $x<-3<2$ (quindi $|2x-4|=-(2x-4)$).
- Per $-3\le x<2$: $x+3\ge0$ (quindi $|x+3|=x+3$), ma $2x-4<0$ ancora, essendo $x<2$ (quindi $|2x-4|=-(2x-4)$).
- Per $x\ge2$: $x+3\ge0$ e $2x-4\ge0$: entrambi i valori assoluti si "aprono" con il segno positivo.

**Passo 4 — sostituiamo in ciascun intervallo:**

- $x<-3$: $-(x+3) - \big(-(2x-4)\big) = -x-3+2x-4 = x-7$.
- $-3\le x<2$: $(x+3) - \big(-(2x-4)\big) = x+3+2x-4 = 3x-1$.
- $x\ge2$: $(x+3) - (2x-4) = x+3-2x+4 = -x+7$.

Il risultato finale, come funzione a tratti:

$$|x+3|-|2x-4| = \begin{cases} x-7 & x<-3 \\ 3x-1 & -3\le x<2 \\ -x+7 & x\ge2 \end{cases}$$

**Perché controlliamo il segno di *entrambi* gli argomenti in *ogni* intervallo, anche quando sembra "ovvio":** un errore comune è concentrarsi solo sul punto critico più vicino e dimenticare di controllare l'altro argomento nello stesso intervallo. Il metodo sistematico (una riga di tabella per ogni intervallo, un controllo per ogni valore assoluto) evita questo tipo di errore, specialmente quando i valori assoluti sono tre o più.

````python
import sympy as sp

x = sp.symbols('x', real=True)
espressione = sp.Abs(x + 3) - sp.Abs(2*x - 4)

# Verifica numerica su un punto di ciascun intervallo
punti_di_prova = {"x<-3": -5, "-3<=x<2": 0, "x>=2": 5}
formule = {"x<-3": lambda x: x - 7, "-3<=x<2": lambda x: 3*x - 1, "x>=2": lambda x: -x + 7}

for zona, x_val in punti_di_prova.items():
    valore_originale = espressione.subs(x, x_val)
    valore_formula = formule[zona](x_val)
    print(f"{zona} (x={x_val}): originale={valore_originale}, formula a tratti={valore_formula}, coincidono: {valore_originale == valore_formula}")
````

## Riepilogo finale

- $|f(x)|$ si può sempre scrivere come funzione a tratti: $f(x)$ dove $f(x)\ge0$, e $-f(x)$ dove $f(x)<0$ — serve prima lo studio del segno di $f(x)$ (notebook 1.9).
- Il grafico di $|f(x)|$ si ottiene da quello di $f(x)$ ribaltando verso l'alto le parti che stanno sotto l'asse $x$.
- Con più valori assoluti, si individuano tutti i punti critici, si ordinano, e si riscrive l'espressione senza valori assoluti in ciascun intervallo così creato.
- Espressioni come $|x-a|+|x-b|$ hanno un'interpretazione geometrica come somma di distanze, e diventano costanti nell'intervallo tra i due punti critici.
