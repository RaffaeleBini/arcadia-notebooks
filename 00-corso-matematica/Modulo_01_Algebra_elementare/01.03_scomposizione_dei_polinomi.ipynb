# 1.3 — Scomposizione dei polinomi

**Prerequisiti:** 1.1 Polinomi (introduzione), 1.2 Prodotti notevoli.


## Perché ci interessa

Nel notebook 0.3 abbiamo imparato a ridurre una frazione numerica ai minimi termini, tipo $\frac{15}{60}=\frac{1}{4}$, trovando i fattori comuni. Con i polinomi vorremmo poter fare la stessa cosa: semplificare $\frac{x^2-1}{x+1}$ richiede di riconoscere che $x^2-1=(x+1)(x-1)$, per poi "cancellare" il fattore comune $(x+1)$.

**Scomporre** (o "fattorizzare") un polinomio significa riscriverlo come prodotto di polinomi più semplici. È l'operazione inversa dello sviluppo con la distributiva (notebook 1.1): invece di espandere un prodotto in una somma, trasformiamo una somma in un prodotto. Ci servirà costantemente per semplificare frazioni algebriche (prossimo notebook) e per risolvere equazioni di grado superiore al primo.

## Teoria, passo dopo passo

### Raccoglimento totale a fattor comune

È la tecnica più semplice: se ogni termine di un polinomio condivide un fattore comune, lo si "estrae" con la distributiva al contrario (esattamente come nel notebook 0.5 e nell'esercizio del notebook 0.1):

$$6x^3+9x^2 = 3x^2(2x+3)$$

Qui il fattore comune è $3x^2$ (il coefficiente 3 divide sia 6 sia 9; $x^2$ è la potenza più piccola di $x$ presente in entrambi i termini).

### Raccoglimento parziale (per gruppi)

Quando non c'è un fattore comune a tutti i termini, a volte si può raggruppare i termini a coppie, raccogliere in ciascun gruppo, e trovare così un fattore comune "di secondo livello":

$$ax+ay+bx+by = a(x+y)+b(x+y) = (x+y)(a+b)$$

Nel primo passaggio abbiamo raccolto $a$ dai primi due termini e $b$ dagli ultimi due; nel secondo, abbiamo notato che $(x+y)$ è ora un fattore comune a entrambi i "blocchi", e lo abbiamo raccolto a sua volta.

### Riconoscere i prodotti notevoli al contrario

Ogni prodotto notevole visto nel notebook 1.2 si può leggere "al contrario", per scomporre:

$$a^2+2ab+b^2 = (a+b)^2 \qquad a^2-2ab+b^2=(a-b)^2 \qquad a^2-b^2=(a+b)(a-b)$$

Esempio: $x^2-9$ è una differenza di quadrati ($x^2-3^2$), quindi $x^2-9=(x+3)(x-3)$. Esempio: $x^2+6x+9$ è il quadrato di un binomio ($x^2+2\cdot x\cdot3+3^2$), quindi $x^2+6x+9=(x+3)^2$.

### Scomposizione del trinomio di secondo grado

Un trinomio $x^2+sx+p$ (coefficiente di $x^2$ uguale a 1) si scompone cercando due numeri $m$ e $n$ tali che:

$$m+n=s \qquad m\cdot n=p$$

Se li troviamo, allora $x^2+sx+p=(x+m)(x+n)$. Verifichiamo perché funziona, sviluppando $(x+m)(x+n)$ con la distributiva: $x^2+nx+mx+mn = x^2+(m+n)x+mn$ — che coincide con $x^2+sx+p$ esattamente quando $m+n=s$ e $mn=p$.

Esempio: scomponi $x^2+5x+6$. Cerchiamo due numeri che sommati diano $5$ e moltiplicati diano $6$: sono $2$ e $3$. Quindi $x^2+5x+6=(x+2)(x+3)$.

### Un polinomio può essere "irriducibile"

Non tutti i polinomi si scompongono in fattori più semplici (con coefficienti razionali). Per esempio $x^2+1$ non si può scomporre in questo modo: non esistono due numeri reali il cui prodotto sia $1$ e la cui somma sia $0$ senza uscire dai numeri reali (vedremo nel Modulo 9, con i numeri complessi, che in realtà si può scomporre come $(x+i)(x-i)$ — ma non con gli strumenti che abbiamo ora).

````python
import sympy as sp

x, a, b, y = sp.symbols('x a b y')

# SymPy scompone automaticamente con factor()
print("6x^3+9x^2 =", sp.factor(6*x**3 + 9*x**2))
print("x^2-9 =", sp.factor(x**2 - 9))
print("x^2+6x+9 =", sp.factor(x**2 + 6*x + 9))
print("x^2+5x+6 =", sp.factor(x**2 + 5*x + 6))

# x^2+1 non si scompone sui razionali/reali
print("\nx^2+1 scomposto sui reali:", sp.factor(x**2 + 1))  # resta invariato
print("x^2+1 scomposto sui complessi:", sp.factor(x**2 + 1, extension=[sp.I]))

# Verifica del raccoglimento parziale
espressione = a*x + a*y + b*x + b*y
print("\nax+ay+bx+by =", sp.factor(espressione))
````

## Esercizio svolto e commentato

**Problema:** scomponi completamente $3x^3-12x$.

**Ragionamento:**

Procediamo per passi, applicando prima la tecnica più semplice (raccoglimento) e poi controllando se il risultato nasconde altri schemi riconoscibili.

**Passo 1 — raccoglimento a fattor comune:** entrambi i termini condividono il fattore $3x$ (il coefficiente 3 divide sia 3 sia 12; $x$ è la potenza più piccola presente in entrambi):

$$3x^3-12x = 3x(x^2-4)$$

**Passo 2 — osserviamo il fattore rimasto, $x^2-4$:** è una differenza di quadrati ($x^2-2^2$), quindi si scompone ulteriormente:

$$x^2-4 = (x+2)(x-2)$$

**Passo 3 — componiamo il risultato finale**, sostituendo:

$$3x^3-12x = 3x(x+2)(x-2)$$

Questa è la scomposizione completa: nessuno dei tre fattori ($3x$, $x+2$, $x-2$) si può scomporre ulteriormente.

**Perché è importante controllare "fino in fondo" dopo ogni raccoglimento:** fermarsi al primo passo ($3x(x^2-4)$) darebbe una scomposizione **incompleta**. La regola pratica è: dopo ogni raccoglimento, guarda sempre se il fattore rimasto nasconde un altro schema riconoscibile (prodotto notevole, altro raccoglimento) prima di considerare il lavoro finito.

````python
import sympy as sp

x = sp.symbols('x')
risultato = sp.factor(3*x**3 - 12*x)
print("Scomposizione completa:", risultato)  # atteso 3*x*(x-2)*(x+2)
````

## Riepilogo finale

- Scomporre un polinomio significa riscriverlo come prodotto: è l'operazione inversa dello sviluppo con la distributiva.
- **Raccoglimento totale**: si estrae il fattore comune a tutti i termini. **Raccoglimento parziale**: si raggruppa a coppie quando non c'è un fattore comune globale.
- I prodotti notevoli (notebook 1.2) si leggono anche "al contrario" per riconoscere scomposizioni: $a^2-b^2=(a+b)(a-b)$, $a^2\pm2ab+b^2=(a\pm b)^2$.
- Il trinomio $x^2+sx+p$ si scompone in $(x+m)(x+n)$ trovando $m,n$ tali che $m+n=s$ e $mn=p$.
- Dopo ogni raccoglimento, controlla sempre se il fattore rimasto nasconde un altro schema: non tutte le scomposizioni sono complete al primo passaggio. Alcuni polinomi (come $x^2+1$) sono irriducibili sui numeri reali.
