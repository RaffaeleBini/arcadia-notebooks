## Perché ci interessa

Finora abbiamo lavorato con numeri: $3$, $\frac{1}{2}$, $\sqrt{2}$. Ma moltissimi problemi (calcolare un'area, descrivere un moto, modellare un profitto) coinvolgono quantità che **non conosciamo ancora**, e che indichiamo con una lettera: $x$.

Quando combiniamo lettere e numeri con addizione, sottrazione e moltiplicazione (mai divisione per una lettera, per ora), otteniamo un **polinomio**. I polinomi sono probabilmente l'oggetto algebrico più usato in tutta la matematica: li ritroveremo in ogni modulo da qui in avanti, dalle equazioni alle funzioni, dai limiti agli integrali.

## Teoria, passo dopo passo

### Dal numero al monomio

Un'espressione come $5x^2$ si chiama **monomio**: è il prodotto di un numero (il **coefficiente**, qui $5$) e una o più lettere elevate a potenza (la **parte letterale**, qui $x^2$). Il **grado** del monomio è la somma degli esponenti della parte letterale: $5x^2$ ha grado 2, $3xy$ ha grado 2 (esponente 1 su $x$, esponente 1 su $y$, $1+1=2$), $7$ (senza lettere) ha grado 0.

### Dal monomio al polinomio

Un **polinomio** è una somma di monomi, chiamati i suoi **termini**:

$$P(x) = 3x^2 - 5x + 7$$

Il grado del polinomio è il grado più alto tra i suoi termini: qui il grado è 2. Un polinomio di grado 1 si chiama **binomio** se ha due termini di grado diverso da zero non combinabili (es. $2x+3$), un polinomio con tre termini si chiama **trinomio**.

Due monomi si dicono **simili** se hanno la stessa parte letterale (stesse lettere, agli stessi esponenti): $3x^2$ e $-5x^2$ sono simili, $3x^2$ e $3x$ no. Solo i monomi simili si possono sommare direttamente, sfruttando la proprietà distributiva (notebook 0.5) al contrario:

$$3x^2 - 5x^2 = (3-5)x^2 = -2x^2$$

### Somma e sottrazione di polinomi

Per sommare due polinomi, si sommano i coefficienti dei termini simili (esattamente il "raccogliere a fattor comune" visto nel notebook 0.5), lasciando invariati i termini non simili:

$$(3x^2-5x+7) + (x^2+2x-3) = (3+1)x^2+(-5+2)x+(7-3) = 4x^2-3x+4$$

Sottrarre un polinomio equivale a sommare il suo opposto (notebook 0.2): si cambia segno a ogni suo termine.

$$(3x^2-5x+7) - (x^2+2x-3) = 3x^2-5x+7 - x^2-2x+3 = 2x^2-7x+10$$

### Moltiplicazione di polinomi

Per moltiplicare due polinomi si usa la proprietà distributiva, applicandola più volte: ogni termine del primo polinomio va moltiplicato per ogni termine del secondo, poi si sommano i risultati (raccogliendo i termini simili).

$$(x+2)(x+3) = x\cdot x + x\cdot 3 + 2\cdot x + 2\cdot 3 = x^2+3x+2x+6 = x^2+5x+6$$

Quando si moltiplicano potenze della stessa lettera, si usano le regole delle potenze viste nel notebook 0.6: $x^2 \cdot x^3 = x^5$.

### Perché il grado si somma nella moltiplicazione

Se $P(x)$ ha grado $m$ e $Q(x)$ ha grado $n$, il prodotto $P(x)\cdot Q(x)$ ha grado $m+n$. Il motivo è diretto: il termine di grado più alto di $P$, moltiplicato per il termine di grado più alto di $Q$, dà un termine di grado $m+n$ (si sommano gli esponenti, proprietà delle potenze), e nessun altro prodotto di termini può raggiungere un grado più alto.

````python
import sympy as sp

x, y = sp.symbols('x y')

# Somma e sottrazione di polinomi
P = 3*x**2 - 5*x + 7
Q = x**2 + 2*x - 3

somma = sp.expand(P + Q)
differenza = sp.expand(P - Q)
print("P + Q =", somma)
print("P - Q =", differenza)

# Moltiplicazione di polinomi: sp.expand esegue la distributiva per esteso
prodotto = sp.expand((x + 2) * (x + 3))
print("\n(x+2)(x+3) =", prodotto)

# Verifica sul grado del prodotto: grado(P*Q) = grado(P) + grado(Q)
grado_P = sp.Poly(P, x).degree()
grado_Q = sp.Poly(Q, x).degree()
grado_prodotto = sp.Poly(sp.expand(P * Q), x).degree()
print(f"\ngrado(P)={grado_P}, grado(Q)={grado_Q}, grado(P*Q)={grado_prodotto}")
````

## Esercizio svolto e commentato

**Problema:** calcola $(2x^2-3x+1) - (x^2-4x+5) + (x-2)(x+1)$, semplificando il risultato.

**Ragionamento:**

Procediamo un pezzo alla volta, rispettando l'ordine delle operazioni (notebook 0.7): prima il prodotto tra parentesi (più "stretto"), poi le somme/sottrazioni tra polinomi.

**Passo 1 — sviluppiamo il prodotto $(x-2)(x+1)$** con la distributiva:

$$(x-2)(x+1) = x\cdot x + x\cdot1 - 2\cdot x - 2\cdot1 = x^2+x-2x-2 = x^2-x-2$$

**Passo 2 — riscriviamo tutta l'espressione** con il prodotto già sviluppato:

$$(2x^2-3x+1) - (x^2-4x+5) + (x^2-x-2)$$

**Passo 3 — eliminiamo la parentesi con il segno meno**, cambiando segno a ogni suo termine (notebook 0.2, sottrazione come somma dell'opposto):

$$2x^2-3x+1 - x^2+4x-5 + x^2-x-2$$

**Passo 4 — raccogliamo i termini simili**, sommando separatamente i coefficienti di $x^2$, di $x$, e i termini noti:

- Termini in $x^2$: $2-1+1=2$, quindi $2x^2$.
- Termini in $x$: $-3+4-1=0$, quindi il termine in $x$ **sparisce**.
- Termini noti: $1-5-2=-6$.

Il risultato finale è:

$$2x^2-6$$

**Perché conviene tenere separati i tre "gruppi" (termini in $x^2$, in $x$, noti) invece di sommare tutto insieme alla rinfusa:** i monomi simili sono le uniche quantità che si possono davvero combinare (proprietà distributiva al contrario). Tenerli organizzati per grado riduce il rischio di dimenticare un termine o di sommare per errore due monomi non simili.

````python
import sympy as sp

x = sp.symbols('x')
espressione = (2*x**2 - 3*x + 1) - (x**2 - 4*x + 5) + (x - 2)*(x + 1)
risultato = sp.expand(espressione)
print("Risultato:", risultato)  # atteso 2*x**2 - 6
````

## Riepilogo finale

- Un **monomio** è il prodotto di un coefficiente numerico e una parte letterale; un **polinomio** è una somma di monomi (i suoi termini).
- Il **grado** di un monomio è la somma degli esponenti della parte letterale; il grado di un polinomio è il grado più alto tra i suoi termini.
- Solo i **monomi simili** (stessa parte letterale) si sommano direttamente, raccogliendo i coefficienti (proprietà distributiva).
- Moltiplicare due polinomi significa applicare la distributiva più volte: ogni termine per ogni termine, poi si sommano i termini simili.
- Il grado del prodotto di due polinomi è la somma dei loro gradi — a meno che un coefficiente non si annulli per un valore particolare di un parametro.
