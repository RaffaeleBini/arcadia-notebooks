```python
import torch.nn as nn

# Una única neurona: una capa Linear que recibe un valor y produce un valor
modelo = nn.Linear(1, 1)

# Un único dato de entrada, de forma (1, 1)
entrada = torch.tensor([[2.0]])

# Forward pass: PyTorch llama a la capa directamente
salida = modelo(entrada)

# El valor real que el modelo debería haber predicho
target = torch.tensor([[5.0]])

# Función de pérdida: error cuadrático medio entre la predicción y el valor real
loss_fn = nn.MSELoss()
loss = loss_fn(salida, target)

# Propagamos el error hacia atrás: PyTorch calcula el gradiente de la pérdida
# respecto a cada parámetro de la capa (peso y sesgo)
loss.backward()

print("salida:", salida)
print("loss:", loss)
print("gradiente del peso:", modelo.weight.grad)
print("gradiente del sesgo:", modelo.bias.grad)
```
