### Passo 2 — Scegli la tecnica del corso che applicherai

Scegli l'architettura in base al tipo di dato di input, non il contrario:

Il tuo notebook deve percorrere, con il tuo problema, le cinque fasi dello schema del corso:

1. Carica i dati e costruiscili come tensori di PyTorch
2. Rappresentali con un `Dataset` e servili in lotti con un `DataLoader`
3. Definisci il tuo modello come una classe che eredita da `nn.Module`
4. Addestra il modello con un ciclo di `forward`, calcolo della perdita, `backward` e `optimizer.step()`, monitorando la perdita dentro e fuori dal campione per rilevare il sovradattamento (lezione 8)
