## Parte 1 — Deep learning e reti neurali

### Deep learning rispetto al machine learning tradizionale

Il machine learning (ML) già inverte la logica della programmazione tradizionale: invece di scrivere regole a mano, dai al sistema esempi e lo lasci dedurre il pattern. Il deep learning (DL) è un sottocampo del ML che utilizza reti neurali con vari strati nascosti impilati, capaci di imparare autonomamente quali caratteristiche (features) dei dati sono importanti, invece di farle definire a mano.

È la differenza tra seguire una ricetta già perfezionata da un'altra persona (ML tradizionale, dove tu scegli le caratteristiche) e lasciare che un apprendista provi migliaia di combinazioni fino a trovare la propria ricetta (DL, dove il modello impara le caratteristiche). Questa capacità è ciò che ha reso il DL la tecnica dominante nella visione artificiale, nell'elaborazione del linguaggio naturale e nel riconoscimento vocale.

### Che cos'è una rete neurale

Una rete neurale (NN) è un modello matematico ispirato, in modo semplificato, a come si connettono i neuroni del cervello. Si organizza in strati:

- **Strato di input**: riceve i dati già convertiti in numeri
- **Strati nascosti**: strati intermedi su cui non hai controllo diretto
- **Strato di output**: produce la previsione finale

Ogni neurone moltiplica i suoi input per alcuni pesi, somma un bias e passa il risultato per una **funzione di attivazione**, che introduce la non linearità necessaria per imparare pattern complessi. Due funzioni comuni sono ReLU (Rectified Linear Unit), che annulla i valori negativi e lascia passare quelli positivi, e sigmoid, che comprime qualsiasi valore in un intervallo tra zero e uno.

Puoi vederne la forma con un po' di codice, anche se non fa ancora parte di nessuna rete:

```
import torch
import matplotlib.pyplot as plt

x = torch.linspace(-5, 5, 100)
relu = torch.relu(x)
sigmoide = torch.sigmoid(x)

plt.figure(figsize=(6, 4))
plt.plot(x, relu, label="ReLU")
plt.plot(x, sigmoide, label="sigmoid")
plt.axhline(0, color="gray", linewidth=0.5)
plt.legend()
plt.title("Dos funciones de activación habituales")
plt.show()
```

## Parte 2 — Tensori e autograd in PyTorch

### Cos'è un tensore

PyTorch rappresenta assolutamente tutto —un numero singolo, un'immagine, un lotto di migliaia di immagini— con un'unica struttura: il **tensore**, un array multidimensionale simile a un `ndarray` di NumPy, con il vantaggio di poter essere eseguito su una GPU (Graphics Processing Unit). Verifichiamolo con il codice.

```
print("Versión de PyTorch:", torch.__version__)

# Un tensor de una dimensión (vector), a partir de una lista de Python
x = torch.tensor([5, 3])
y = torch.tensor([2, 1])

# PyTorch sobrecarga los operadores habituales: esto multiplica elemento a elemento
print("x * y =", x * y)
```

```
# torch.zeros crea un tensor lleno de ceros con la forma (shape) que le indiques
z = torch.zeros((2, 5))
print("z:\n", z)
print("shape:", z.shape)

# view reorganiza los mismos datos en una forma distinta, sin copiarlos
z_reordenado = z.view(1, 10)
print("shape tras view(1, 10):", z_reordenado.shape)
```

## Parte 2 — Tensor e autograd in PyTorch

### Che cos'è un tensor

PyTorch rappresenta assolutamente tutto — un numero singolo, un'immagine, un batch di migliaia di immagini — con una sola struttura: il **tensor**, un array multidimensionale simile a un `ndarray` di NumPy, con il vantaggio di poter essere eseguito su una GPU (Graphics Processing Unit). Vediamo di verificarlo con il codice.

```
# requires_grad=True le dice a PyTorch que registre cada operación sobre x
x = torch.tensor([2.0, 3.0], requires_grad=True)

y = x + 2
z = y * y * 3

# backward() recorre el grafo hacia atrás y calcula el gradiente de z respecto a x
z.backward(torch.tensor([1.0, 1.0]))

print("x.grad =", x.grad)
```

Non devi risolvere questa derivata a mano per fidarti del risultato: l'importante è che `.backward()` attivi tutto il calcolo dei gradienti e che `x.grad` sia dove PyTorch li rende disponibili.

### Una rete neurale minima con `nn.Module` e `nn.Linear`

Ogni rete neurale in PyTorch segue lo stesso schema: una classe che eredita da `nn.Module`, con i suoi strati dichiarati in `__init__` e la logica del forward pass nel metodo `forward`.

```
import torch.nn as nn

class RedMinima(nn.Module):
    def __init__(self):
        super().__init__()
        # nn.Linear(3, 1): una capa totalmente conectada que recibe tres
        # valores de entrada y produce un único valor de salida
        self.capa = nn.Linear(3, 1)

    def forward(self, x):
        return self.capa(x)

modelo = RedMinima()
print(modelo)
```

```
# Un único ejemplo de entrada, con tres características
entrada = torch.tensor([[1.0, 2.0, 3.0]])

# El forward pass: PyTorch llama a forward() automáticamente al usar modelo(entrada)
salida = modelo(entrada)
print("Salida del modelo:", salida)
```

```
# Como salida depende de pesos con requires_grad=True (por defecto en nn.Linear),
# podemos propagar el error hacia atrás y obtener el gradiente de cada peso
salida.backward()

print("Gradiente de los pesos:", modelo.capa.weight.grad)
print("Gradiente del sesgo:", modelo.capa.bias.grad)
```

## Riepilogo della lezione

- Il deep learning è un sottocampo del ML che usa reti con più strati per imparare rappresentazioni dei dati, senza che tu definisca a mano quali caratteristiche contano
- Una rete neurale si organizza in strati di input, nascosti e output; ogni connessione ha un peso, e una funzione di attivazione introduce la non linearità necessaria per imparare pattern complessi
- Un tensore è la struttura centrale di PyTorch: un array multidimensionale che rappresenta qualsiasi dato e può essere eseguito su CPU o su GPU
- Autograd calcola automaticamente i gradienti dei tensori con `requires_grad=True` quando chiami `.backward()`
- In PyTorch, una rete si definisce come una classe che eredita da `nn.Module`, con strati come `nn.Linear` in `__init__` e la logica del forward pass in `forward`
