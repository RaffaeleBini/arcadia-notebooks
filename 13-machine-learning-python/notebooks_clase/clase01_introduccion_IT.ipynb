## ¿Qué es el machine learning?

En 1959, Arthur Samuel definió el ML como el campo que da a las máquinas la capacidad de aprender sin programarlas explícitamente para cada tarea. Es una rama de la inteligencia artificial (IA).

¿Por qué no basta con programar reglas fijas? Piensa en el spam: ¿cuántas reglas escritas a mano harían falta para detectar todos los correos basura del mundo? Con el ML no le dictas las reglas al programa una por una: le muestras muchos esempios (datos) y el propio algoritmo ajusta i suoi parametri interni hasta encontrar un patrón útil.

Es exactamente lo que hace un niño o una niña aprendiendo a reconocer perros: nadie le da una lista de reglas exactas, ve muchos esempios y generaliza.

## Los tres tipos de aprendizaje

- **Aprendizaje supervisado**: aprendes a partire da esempios etiquetados, es decir, casos donde ya conoces la respuesta correcta. Por esempio, fotos de frutas ya clasificadas como "madura" o "verde"
- **Aprendizaje no supervisado**: el modelo busca patrones en datos sin etiquetas, como agrupar clientes con hábitos de compra similes sin decirle de antemano qué grupos existen
- **Aprendizaje por refuerzo**: el modelo aprende interactuando con un entorno y recibiendo una señal de recompensa, ajustando su comportamiento para maximizarla

En este curso nos vamos a centrar sobre todo en aprendizaje supervisado (regresión, KNN, SVM) y en un poco de aprendizaje no supervisado (agrupamiento) en los últimos capítulos.

## Il flusso di un progetto di ML

Quasi tutti i progetti di ML seguono lo stesso flusso, qualunque sia l'algoritmo che scegli alla fine. Conoscere questa mappa da subito ti evita di perderti più avanti.

1. **Preprocessare i dati**: i dati reali quasi non arrivano mai puliti (valori mancanti, testo dove ti aspetti numeri, scale diverse). Qui pulisci e trasformi
2. **Scegliere l'algoritmo**: in base al problema (prevedere un numero, classificare in categorie, raggruppare elementi), scegli regressione, KNN, SVM o un algoritmo di clustering
3. **Addestrare il modello**: dividi i dati in un insieme di addestramento e uno di test, e il modello apprende solo con quello di addestramento
4. **Valutare le prestazioni**: usi l'insieme di test, che il modello non ha mai visto, per verificare se generalizza bene
5. **Prevedere con dati nuovi**: usi il modello già addestrato su casi che non ha mai visto

**Domanda per il gruppo:** perché pensi che separiamo i dati in "addestramento" e "test" invece di usare tutti i dati per addestrare? Pensa a come studi per un esame: ripasseresti solo con le domande che sai già che usciranno?

## Un primer vistazo a un dataset

Todavía no vamos a entrenar ningún modelo, eso empieza en la clase 2. Pero antes de seguir conviene que veas de un vistazo qué aspecto tiene un dataset (insieme di dati) en Python.

scikit-learn incluye varios datasets de juguete listos para practicar. Vamos a cargar el clásico dataset iris, que recoge medidas de flores de tres especies distintas.

**Pregunta para el grupo:** al ver la tabla, ¿qué crees que representa cada fila? ¿Y cada columna?

```python
from sklearn.datasets import load_iris

# Cargamos el dataset de juguete iris
iris = load_iris(as_frame=True)

# scikit-learn nos da los datos ya organizados como un DataFrame de pandas
df_iris = iris.frame

# Mostramos las primeras filas para ver qué aspecto tiene
df_iris.head()
```

Vamos a ver también un resumen estadístico rápido con `describe()`, una herramienta de pandas muy útil para hacerte una idea general de tus datos antes de tocar nada.

```python
# describe() te da un resumen estadístico rápido de cada columna numérica
df_iris.describe()
```
