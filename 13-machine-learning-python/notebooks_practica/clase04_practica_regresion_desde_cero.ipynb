```python
    # La pendiente sale de las medias de xs, ys y de su producto,
    # respetando el orden de las operaciones (paréntesis primero)

        ((np.mean(xs) * np.mean(ys)) - np.mean(xs * ys)) /

    b = np.mean(ys) - m * np.mean(xs)
```

```python
def squared_error(ys_orig, ys_line):
    # Suma de las diferencias al cuadrado entre la línea y los valores reales
    return sum((ys_line - ys_orig) ** 2)


def coefficient_of_determination(ys_orig, ys_line):
    # Línea horizontal en la media de Y, con un valor por cada punto original
    y_mean_line = [np.mean(ys_orig) for _ in ys_orig]

    # Error de tu recta frente al error de la línea de la media
    squared_error_regr = squared_error(ys_orig, ys_line)
    squared_error_y_mean = squared_error(ys_orig, y_mean_line)

    # R cuadrado: uno menos la proporción de error que le queda a tu recta
    return 1 - (squared_error_regr / squared_error_y_mean)


r_squared = coefficient_of_determination(ys, regression_line)
print("R cuadrado (implementación propia):", r_squared)
```
