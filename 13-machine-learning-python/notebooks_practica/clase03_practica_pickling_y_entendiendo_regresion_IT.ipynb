## Ejercicio 1: guarda un modello y comprueba que predice igual tras cargarlo

Vas a simular un conjunto de dati distinto al de la lezione: la relación entre las horas que alguien dedica a estudiar y la nota que obtiene en un examen. Ya tienes el código que genera los dati y entrena el modello. Tu tarea es completar el guardado y la carga con `pickle`.

**Suggerimenti:**

- Para guardar, abre un archivo en modo `"wb"` con `open(...)` dentro de un `with`, y usa `pickle.dump(objeto, archivo)`
- Para simular una nueva sesión, primero verás una cella que borra la variabile del modello original con `del`
- Para cargar, abre el archivo en modo `"rb"` y usa `modello_cargado = pickle.load(archivo)`
- Al final, compara la predicción del modello original (`prediccion_referencia`) con la del modello cargado (`prediccion_nueva`): deben coincidir exactamente

````python
import numpy as np
from sklearn.linear_model import LinearRegression

# Simulamos un conjunto de datos distinto: horas de estudio frente a nota de examen
np.random.seed(11)
horas_estudio = np.arange(1, 61).reshape(-1, 1)
nota_examen = 2.5 * horas_estudio.flatten() + np.random.normal(0, 4, size=60)

modelo_notas = LinearRegression()
modelo_notas.fit(horas_estudio, nota_examen)

prediccion_referencia = modelo_notas.predict([[30]])
print("Predicción antes de guardar (30 horas de estudio):", prediccion_referencia)
````

## Ejercicio 2: escala colonnas con rangos muy distintos e interpreta el risultato

Ahora simulas dati de viviendas con dos feature en escalas muy diferentes: la superficie en metros cuadrados (entre 40 y 300, aproximadamente) y la antigüedad en años (entre 0 y 50). Vas a aplicar `StandardScaler` y comprobar cómo cambia cada colonna.

**Suggerimenti:**

- Crea el escalador con `StandardScaler()`
- Ajusta y transforma los dati de addestramento en un solo paso con `fit_transform()`
- Para los dati de prueba, usa solo `transform()` (sin `fit`), para no filtrar información del conjunto de prueba
- Después de escalar, cada colonna debe tener media aproximada a cero y desviación estándar aproximada a uno
- Para interpretar el risultato, compara `X_train.mean(axis=0)` y `X_train.std(axis=0)` antes y después de escalar

````python
import numpy as np
from sklearn.model_selection import train_test_split

# Simulamos datos de viviendas: superficie (m2) y antigüedad (años), en escalas muy distintas
np.random.seed(23)
superficie = np.random.uniform(40, 300, size=80)
antiguedad = np.random.uniform(0, 50, size=80)
precio_vivienda = 800 * superficie - 500 * antiguedad + np.random.normal(0, 5000, size=80)

X_viviendas = np.column_stack([superficie, antiguedad])
y_viviendas = precio_vivienda

X_train, X_test, y_train, y_test = train_test_split(
    X_viviendas, y_viviendas, test_size=0.2, random_state=23
)

print("Sin escalar, primera fila de X_train:", X_train[0])
````

```python
# TODO 1: importar StandardScaler
from sklearn.preprocessing import StandardScaler

# TODO 2: crear el escalador
escalador_viviendas = StandardScaler()

# TODO 3: ajustar y transformar los dati de addestramento
X_train_escalado = escalador_viviendas.fit_transform(X_train)

# TODO 4: transformar los dati de prueba con la misma transformación aprendida
X_test_escalado = escalador_viviendas.transform(X_test)

# TODO 5: comparar medias y desviaciones antes y después de escalar
print("--- Antes de escalar ---")
print("Media por colonna:", X_train.mean(axis=0))
print("Desviación estándar por colonna:", X_train.std(axis=0))

print("\n--- Después de escalar ---")
print("Media por colonna:", X_train_escalado.mean(axis=0).round(2))
print("Desviación estándar por colonna:", X_train_escalado.std(axis=0).round(2))

# TODO 6:
# Antes de escalar, la colonna "superficie" tiene valorees de decenas o centenas
# (entre 40 y 300), mientras que "antigüedad" va de 0 a 50. Esa diferencia de
# magnitud hace que, para un algoritmo que compara distancias o magnitudes numéricas,
# la superficie parezca "más importante" solo por tener números más grandes, aunque
# la antigüedad también influya en el precio. Tras escalar, ambas colonnas quedan
# con media cero y desviación estándar uno, así que se vuelven comparables entre sí.
```

Con esta comparación ves de forma directa lo que explica la lezione: `StandardScaler` no cambia la relación entre tus dati, solo los reescala para que ninguna colonna domine el aprendizaje por el simple hecho de tener números más grandes.
