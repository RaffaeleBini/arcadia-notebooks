## Ejercicio 1: guarda un modelo y comprueba que predice igual tras cargarlo

Vas a simular un conjunto de datos distinto al de la clase: la relación entre las horas que alguien dedica a estudiar y la nota que obtiene en un examen. Ya tienes el código que genera los datos y entrena el modelo. Tu tarea es completar el guardado y la carga con `pickle`.

**Pistas:**

- Para guardar, abre un archivo en modo `"wb"` con `open(...)` dentro de un `with`, y usa `pickle.dump(objeto, archivo)`
- Para simular una nueva sesión, primero verás una celda que borra la variable del modelo original con `del`
- Para cargar, abre el archivo en modo `"rb"` y usa `modelo_cargado = pickle.load(archivo)`
- Al final, compara la predicción del modelo original (`prediccion_referencia`) con la del modelo cargado (`prediccion_nueva`): deben coincidir exactamente

````python
import numpy as np
from sklearn.linear_model import LinearRegression

# Simulamos un conjunto de datos distinto: horas de estudio frente a nota de examen
np.random.seed(11)
horas_estudio = np.arange(1, 61).reshape(-1, 1)
nota_examen = 2.5 * horas_estudio.flatten() + np.random.normal(0, 4, size=60)

modelo_notas = LinearRegression()
modelo_notas.fit(horas_estudio, nota_examen)

prediccion_referencia = modelo_notas.predict([[30]])
print("Predicción antes de guardar (30 horas de estudio):", prediccion_referencia)
````

## Ejercicio 2: escala columnas con rangos muy distintos e interpreta el resultado

Ahora simulas datos de viviendas con dos características en escalas muy diferentes: la superficie en metros cuadrados (entre 40 y 300, aproximadamente) y la antigüedad en años (entre 0 y 50). Vas a aplicar `StandardScaler` y comprobar cómo cambia cada columna.

**Pistas:**

- Crea el escalador con `StandardScaler()`
- Ajusta y transforma los datos de entrenamiento en un solo paso con `fit_transform()`
- Para los datos de prueba, usa solo `transform()` (sin `fit`), para no filtrar información del conjunto de prueba
- Después de escalar, cada columna debe tener media aproximada a cero y desviación estándar aproximada a uno
- Para interpretar el resultado, compara `X_train.mean(axis=0)` y `X_train.std(axis=0)` antes y después de escalar

````python
import numpy as np
from sklearn.model_selection import train_test_split

# Simulamos datos de viviendas: superficie (m2) y antigüedad (años), en escalas muy distintas
np.random.seed(23)
superficie = np.random.uniform(40, 300, size=80)
antiguedad = np.random.uniform(0, 50, size=80)
precio_vivienda = 800 * superficie - 500 * antiguedad + np.random.normal(0, 5000, size=80)

X_viviendas = np.column_stack([superficie, antiguedad])
y_viviendas = precio_vivienda

X_train, X_test, y_train, y_test = train_test_split(
    X_viviendas, y_viviendas, test_size=0.2, random_state=23
)

print("Sin escalar, primera fila de X_train:", X_train[0])
````
