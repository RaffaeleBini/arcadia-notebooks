```python
valores_k = range(1, 21)
precisiones_vino = []

for k in valores_k:
    clasificador_k = KNeighborsClassifier(n_neighbors=k)
    clasificador_k.fit(X_train, y_train)
    precision_k = clasificador_k.score(X_test, y_test)

for k, precision_k in zip(valores_k, precisiones_vino):
    print(f"K = {k} -> precisión = {precision_k:.4f}")

# Buscamos la posición de la mejor precisión y el K que le corresponde
mejor_precision = max(precisiones_vino)
mejor_k = list(valores_k)[precisiones_vino.index(mejor_precision)]

print("\nMejor K encontrado:", mejor_k, "con precisión:", mejor_precision)
```
