## ¿Qué es el aprendizaje por refuerzo?

Piensa en cómo entrenas a una mascota: no le entregas un manual de movimientos correctos, le das una golosina cuando hace lo que quieres. Poco a poco, asocia la acción con la recompensa. Así funciona el RL.

- El **agente** interactúa con un entorno, prueba acciones y recibe una señal que le indica si esa acción fue buena o mala
- Con el tiempo, aprende a elegir las acciones que maximizan esa señal
- No existe un conjunto de respuestas correctas como en el aprendizaje supervisado: nadie le dice "esta era la acción correcta", el agente lo descubre por ensayo y error

**El RL ya está detrás de logros muy concretos**: AlphaGo derrotando a campeones de Go, robots que manipulan objetos sin dañarlos, estrategias de negociación automática en mercados financieros, coches autónomos y sistemas de recomendación

> 🎤 **Guion de clase**: pregunta a la clase si conocen algún caso de uso de IA en su empresa o sector donde nadie pueda escribir de antemano "la respuesta correcta" para cada situación. Esa es la señal de que RL podría encajar.


## Los tres protagonistas: agente, entorno y recompensa

Todo problema de RL se describe con tres elementos y un bucle que se repite una y otra vez:

- **Agente**: quien toma las decisiones (un robot, un programa de trading, un personaje de videojuego)
- **Entorno**: todo aquello con lo que el agente interactúa (un almacén, un mercado financiero, un tablero de juego)
- **Recompensa**: un número que el entorno devuelve tras cada acción, e indica si fue beneficiosa o perjudicial

**El bucle**: el agente observa el **estado** actual, elige una **acción**, el entorno cambia a un nuevo estado y entrega una **recompensa**. Se repite una y otra vez. A la secuencia completa de estados, acciones y recompensas de un agente la llamamos **episodio** o **trayectoria**.

> 🎤 **Guion de clase**: dibuja en la pizarra el ciclo estado → acción → recompensa → nuevo estado antes de pasar al código. Es el diagrama que vais a ver reflejado línea por línea dentro de un rato.


## El diseño de la recompensa importa (mucho)

Un robot que recibe recompensa positiva por acercarse a un objetivo y negativa por chocar puede encontrar "atajos" que no querías: aprende exactamente lo que le pagas por aprender, no necesariamente lo que tú querías enseñarle.

- Diseña mal la recompensa y el comportamiento final se tuerce, por buena que sea tu intención
- A veces conviene sacrificar una ganancia inmediata a cambio de un beneficio mayor más adelante
- El **factor de descuento** (γ, gamma) es un valor entre cero y uno que indica cuánto peso das a las recompensas futuras frente a las inmediatas
- Un γ cercano a uno da casi la misma importancia al futuro que al presente; un γ cercano a cero hace que el agente solo piense en el corto plazo

> 🎤 **Guion de clase**: pide un ejemplo de la vida cotidiana donde renunciar a un beneficio inmediato traiga una recompensa mayor después (ahorrar en vez de gastar, estudiar en vez de salir). Ese es exactamente el papel de γ.


## La política: cómo decide el agente

La estrategia que sigue el agente para elegir acciones se llama **política** y se representa con la letra griega pi (π). Existen dos tipos:

- **Determinista**: para cada estado, siempre elige la misma acción ("en este cruce, gira siempre a la izquierda")
- **Estocástica**: para cada estado, asigna una probabilidad a cada acción posible ("en este cruce, giro a la izquierda ocho de cada diez veces")

En un laberinto, una política determinista que siempre gira a la izquierda funciona bien si el laberinto es fijo, pero se queda atascada ante un obstáculo inesperado. Una política estocástica, al variar de vez en cuando, tiene más opciones de descubrir atajos nuevos o de adaptarse a cambios. Esa idea conecta directamente con el dilema exploración/explotación que veréis en la próxima clase.

El objetivo final del aprendizaje es encontrar la **política óptima** (π\*): la que maximiza la recompensa acumulada esperada a largo plazo.


## El formalismo detrás del entorno: el PDM

Para programar y analizar un problema de RL, describimos el entorno con un modelo matemático llamado Proceso de Decisión de Markov (PDM). Hoy solo necesitas sus cuatro piezas:

- **S**: el conjunto de todos los estados posibles del entorno
- **A**: el conjunto de todas las acciones que el agente puede ejecutar
- **P**: la función de transición, que indica la probabilidad de pasar de un estado a otro al ejecutar una acción
- **R**: la función de recompensa, que indica qué recompensa recibe el agente en cada caso

La "propiedad de Markov" dice que, para decidir qué pasa a continuación, solo importa el estado actual, no toda la historia anterior. Es como el ajedrez: para decidir tu próxima jugada, solo cuenta la posición actual del tablero.

> 🎤 **Guion de clase**: en la clase de código de dentro de un momento, S serán las cinco posiciones del pasillo, A serán las dos acciones (izquierda y derecha), y P y R estarán implementadas dentro del método `step`. Señálalo explícitamente cuando lleguéis ahí.


## Una primera idea de Q-learning

Para decidir qué acción tomar, ayuda saber "cuánta recompensa espero conseguir si hago esto". A esa estimación la llamamos **valor de acción** o **valor Q**. Uno de los algoritmos más conocidos para aprenderlo es **Q-learning**:

- El agente mantiene una tabla con un valor estimado para cada combinación de estado y acción
- Cada vez que prueba una acción y observa la recompensa real, ajusta un poco esa estimación
- Con suficientes repeticiones, los valores reflejan con fidelidad cuánta recompensa aporta cada acción en cada estado

Q-learning es "sin modelo" (*model-free*): no necesita conocer de antemano cómo funciona el entorno, aprende solo de la experiencia directa. El enfoque contrario, "basado en modelo" (*model-based*), primero construye una representación del entorno y planifica con ella, como un jugador de ajedrez que simula mentalmente varias jugadas antes de mover una ficha.


## El dilema exploración/explotación, en una frase

Para aprender bien, el agente necesita **explorar** (probar acciones nuevas, por si esconden una recompensa mejor) y también **explotar** (aprovechar lo que ya sabe, para no desperdiciar oportunidades).

- Demasiada exploración: pierde tiempo probando opciones mediocres
- Demasiada explotación: se queda atascado en una solución aceptable sin descubrir otra mejor

Piensa en una aspiradora robótica: si solo limpia las zonas que ya conoce, nunca descubre zonas sucias nuevas; si solo explora zonas nuevas, deja sin limpiar lo que ya sabía que necesitaba atención. Le dedicamos la próxima clase entera a este equilibrio, así que hoy basta con que lo tengas presente.


## Manos a la obra: tu primer entorno en Python

Vas a construir, paso a paso, un entorno mínimo: un **pasillo de cinco celdas**. El agente empieza en la celda 0 (izquierda) y la meta está en la celda 4 (derecha). Puede moverse a la izquierda o a la derecha.

```
[0] [1] [2] [3] [4=META]
```

Todavía no va a aprender nada: el objetivo de hoy es ver el bucle agente-entorno funcionando en código real, celda a celda. Solo necesitas `numpy`, que ya viene preinstalado en Google Colab.

> 🎤 **Guion de clase**: ejecuta cada celda de código en directo, una a una, y pausa antes de correr la siguiente para preguntar "¿qué creéis que va a pasar?". No adelantes resultados.


````python
import numpy as np
````

### El entorno: `PasilloEntorno`

Esta clase implementa el PDM del pasillo: `reset` fija el estado inicial (S) y `step` aplica una acción (A), calcula el nuevo estado (P) y la recompensa (R).

Fíjate en el diseño de la recompensa: cada paso "cuesta" -1, para animar al agente a llegar cuanto antes; al llegar a la meta, recompensa extra de +10.

> 🎤 **Guion de clase**: antes de ejecutar, pregunta "¿qué método creéis que se llama para reiniciar el entorno al principio de un episodio?". Enlaza la respuesta con `reset`.


````python
class PasilloEntorno:
    def __init__(self, longitud=5):
        self.longitud = longitud
        self.meta = longitud - 1
        self.estado = 0  # posición inicial del agente

    def reset(self):
        """Reinicia el entorno y devuelve el estado inicial."""
        self.estado = 0
        return self.estado

    def step(self, accion):
        """
        Ejecuta una acción en el entorno.
        accion: 0 = mover a la izquierda, 1 = mover a la derecha
        Devuelve: (nuevo_estado, recompensa, terminado)
        """
        if accion == 1:
            self.estado = min(self.estado + 1, self.longitud - 1)
        elif accion == 0:
            self.estado = max(self.estado - 1, 0)

        terminado = self.estado == self.meta

        # Diseño de la recompensa: cada paso "cuesta" un poco (-1),
        # para animar al agente a llegar a la meta cuanto antes.
        # Al llegar a la meta, recompensa extra de +10.
        if terminado:
            recompensa = 10
        else:
            recompensa = -1

        return self.estado, recompensa, terminado
````

**Ejecuta la celda anterior** para crear la clase (todavía no pasa nada visible: solo defines el molde). Ahora vas a crear una instancia y a comprobar `reset`.

> 🎤 **Guion de clase**: ejecuta esta celda y pregunta a la clase qué valor esperan ver impreso antes de correrla.


````python
entorno = PasilloEntorno()
estado_inicial = entorno.reset()
print("Estado inicial:", estado_inicial)
````

Prueba ahora un solo `step`, a mano, moviendo el agente a la derecha (acción 1).

> 🎤 **Guion de clase**: pregunta "¿en qué celda va a terminar el agente y qué recompensa esperáis?" antes de ejecutar.


````python
nuevo_estado, recompensa, terminado = entorno.step(1)
print(f"nuevo_estado={nuevo_estado}, recompensa={recompensa}, terminado={terminado}")
````

### La política: un agente que decide al azar

Una política estocástica muy simple: en cada estado, elige la acción al azar, con la misma probabilidad para cada una. No usa ninguna información del estado todavía; en próximas clases la sustituirás por una que sí aprende.


````python
def politica_aleatoria(estado, num_acciones=2):
    return np.random.randint(num_acciones)


accion_ejemplo = politica_aleatoria(estado_inicial)
print("Acción elegida por la política aleatoria:", accion_ejemplo)
````

### El bucle agente-entorno

Aquí se ve el ciclo central del RL: observar estado, elegir acción, recibir nuevo estado y recompensa, y repetir hasta el final del episodio.

> 🎤 **Guion de clase**: antes de ejecutar la función, repasa en voz alta cada línea del `for` y relaciónala con el diagrama de la pizarra (estado → acción → nuevo estado → recompensa).


````python
def ejecutar_episodio(entorno, max_pasos=20, semilla=None):
    if semilla is not None:
        np.random.seed(semilla)

    estado = entorno.reset()
    recompensa_acumulada = 0
    nombres_acciones = {0: "izquierda", 1: "derecha"}

    print(f"Estado inicial: {estado}")

    for paso in range(1, max_pasos + 1):
        accion = politica_aleatoria(estado)
        nuevo_estado, recompensa, terminado = entorno.step(accion)
        recompensa_acumulada += recompensa

        print(
            f"Paso {paso}: estado={estado} -> "
            f"acción={nombres_acciones[accion]} -> "
            f"nuevo_estado={nuevo_estado}, recompensa={recompensa}"
        )

        estado = nuevo_estado

        if terminado:
            print(f"\n¡Meta alcanzada en el paso {paso}!")
            break

    print(f"Recompensa acumulada del episodio: {recompensa_acumulada}")
    return recompensa_acumulada
````

**Ejecuta esta celda ahora** y pregunta a la clase cuántos pasos creen que va a tardar el agente en llegar a la meta. Usamos una semilla fija (`semilla=42`) para que el resultado sea el mismo cada vez que la ejecutéis.

> 🎤 **Guion de clase**: deja que salgan varias apuestas antes de ejecutar. Después de ver el resultado, pregunta por qué el agente a veces retrocede: no tiene ninguna razón para preferir "derecha", porque todavía no ha aprendido nada.


````python
ejecutar_episodio(PasilloEntorno(), max_pasos=20, semilla=42)
````

**Cambia la semilla y vuelve a ejecutar**. Verás una trayectoria distinta, porque cambia la secuencia de acciones aleatorias.


````python
ejecutar_episodio(PasilloEntorno(), max_pasos=20, semilla=7)
````

### ¿Cuánto varía el agente aleatorio de un episodio a otro?

Ejecuta varios episodios seguidos, sin fijar la semilla, y observa cuánto varía el número de pasos necesarios para llegar a la meta.

> 🎤 **Guion de clase**: apunta en la pizarra la recompensa acumulada de cada episodio a medida que se imprime. Cierra la clase señalando que esa variabilidad es justo lo que un algoritmo de aprendizaje, como Q-learning, está diseñado para reducir con el tiempo: en la práctica guiada de hoy comparáis esta política aleatoria contra una que sí usa la información del entorno.


````python
resultados = []
for i in range(5):
    print(f"\n--- Episodio {i + 1} ---")
    r = ejecutar_episodio(PasilloEntorno(), max_pasos=30)
    resultados.append(r)

print("\nRecompensas acumuladas de los 5 episodios:", resultados)
print("Media de recompensa acumulada:", np.mean(resultados))
````

## Resumen de puntos clave

- El aprendizaje por refuerzo (RL) enseña a un agente a decidir mediante prueba y error, sin un conjunto de respuestas correctas como en el aprendizaje supervisado
- El bucle central del RL conecta tres piezas: agente, entorno y recompensa, en un ciclo de estado → acción → nuevo estado → recompensa
- La política (π) es la estrategia que usa el agente para elegir acciones; puede ser determinista o estocástica
- El Proceso de Decisión de Markov (PDM) formaliza el entorno con cuatro elementos: estados (S), acciones (A), transiciones (P) y recompensas (R)
- El factor de descuento (γ) regula cuánto peso das a las recompensas futuras frente a las inmediatas
- Q-learning estima, para cada combinación de estado y acción, cuánta recompensa cabe esperar, y ajusta esa estimación con la experiencia
- El diseño de la recompensa importa muchísimo: una recompensa mal pensada enseña al agente el comportamiento equivocado
- Explorar y explotar son dos necesidades en tensión constante; encontrar el equilibrio adecuado es uno de los grandes retos del RL
