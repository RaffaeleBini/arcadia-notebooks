## Da "posso valutarla" a "posso trovare la migliore"

Nella lezione precedente hai calcolato Vπ, la funzione di valore di una politica π fissa, tramite valutazione iterativa: hai applicato l'equazione di Bellman più volte, passata dopo passata, finché i valori hanno smesso di cambiare. È utile, ma incompleto. Sapere quanto vale seguire una politica concreta non ti dice se quella politica è buona, così come avere una mappa delle distanze dal tuo ufficio a ogni punto della città non ti dice se il tuo percorso abituale è il migliore possibile.

Oggi fai il passo che manca: passi da "posso valutare una politica" a "posso trovare la migliore politica possibile". L'insieme di strumenti che userai si chiama **programmazione dinamica**: una famiglia di algoritmi che calcolano politiche ottimali quando conosci il modello completo dell'ambiente —le probabilità di transizione P e la funzione di ricompensa R—, senza bisogno di interagire con esso per tentativi ed errori.

- Programmare l'iterazione dei valori: fondere valutazione e miglioramento in un unico passaggio

## 1. Dalla valutazione al miglioramento

Immagina di aver già calcolato Vπ per una politica π qualsiasi, per esempio la politica casuale della lezione precedente. Ora fermati su ogni casella del GridWorld e poniti una domanda diversa: "se invece di seguire la politica attuale prendo la migliore azione possibile in questo solo passo, e poi torno a seguire π, miglioro il mio risultato?".

Confrontare questa opzione con ciò che faceva la politica originale e scegliere sempre la migliore azione si chiama **miglioramento delle politiche**: costruire una nuova politica π' che agisce in modo goloso (*greedy*) rispetto a Vπ, cioè che in ogni stato sceglie l'azione che massimizza la ricompensa immediata più il valore scontato dello stato successivo.

Questo quasi mai peggiora le cose. Se in uno stato l'azione golosa coincide con quella che già facevi, non perdi nulla; se è diversa, migliori almeno un po'. Per questo π' è sempre uguale o migliore di π in tutti gli stati —lo stesso principio che applichi quando rivedi il percorso abituale per il lavoro con le informazioni sul traffico di oggi: nel peggiore dei casi resti uguale, nel migliore trovi una vera scorciatoia—.


## 2. Iterazione delle politiche: valutare e migliorare, ancora e ancora

Se migliorare una politica una volta già aiuta, cosa succede se ripeti il ciclo completo? Valuti la politica attuale fino in fondo (calcoli il suo Vπ esatto con la valutazione iterativa), la migliori in modo goloso, valuti la nuova politica, la migliori di nuovo, e così via. Questa procedura si chiama **iterazione delle politiche** (*policy iteration*).

Pensa a due frecce che si inseguono. La valutazione è una freccia verso l'alto: regola i valori finché si adattano perfettamente alla politica attuale. Il miglioramento è una freccia verso il basso: appena hai quei valori esatti, cambi la politica, quindi i valori che hai appena calcolato non descrivono più bene la nuova politica. Il processo ripete questo andirivieni finché, in un ciclo, la politica non cambia più.

A quel punto, l'azione golosa in ogni stato coincide esattamente con ciò che la politica già faceva, il che può significare una sola cosa: vale l'equazione di ottimalità di Bellman, e quindi la politica attuale è già la politica ottimale π\*. Poiché il numero di politiche deterministiche possibili in un PDM (Processo Decisionale di Markov) finito è limitato, e ogni ciclo produce una politica uguale o migliore della precedente, l'algoritmo è garantito nel raggiungere π\* in un numero finito di cicli.

> 🎙️ **Traccia della lezione:** disegna le due frecce alla lavagna (valutazione verso l'alto, miglioramento verso il basso) e sottolinea il criterio di arresto: il ciclo termina quando la politica non cambia più tra un ciclo e il successivo, non quando i valori smettono di cambiare.


## 3. Iterazione dei valori: la scorciatoia diretta

L'iterazione delle politiche funziona, ma ha un punto debole: a ogni ciclo, valutare la politica a fondo —ripetere la valutazione iterativa finché converge del tutto— può costare molti passaggi, e gran parte di quella precisione viene buttata via non appena migliori comunque la politica. Serve davvero aspettare che la valutazione converga completamente prima di migliorare?

La risposta è no, e porta direttamente all'**iterazione dei valori** (*value iteration*): invece di completare la valutazione, la interrompi a un solo passaggio e mescoli valutazione e miglioramento nello stesso passo. In ogni passaggio, ogni stato aggiorna il proprio valore prendendo direttamente il massimo tra le sue azioni —invece di fare la media secondo una politica fissa—, usando l'equazione di ottimalità di Bellman come regola di aggiornamento. Quando questi valori smettono di cambiare in modo apprezzabile, ricavi la politica ottimale con un ultimo sguardo a un passo: in ogni stato scegli l'azione che massimizza ricompensa più valore scontato dello stato successivo.

Puoi vederlo come un caso particolare della stessa famiglia dell'iterazione delle politiche: è un'iterazione delle politiche con la fase di valutazione ridotta a un solo passaggio. Di solito richiede più passaggi singoli di quanti cicli completi richieda l'iterazione delle politiche, ma ogni passaggio è molto più economico, quindi il costo totale finisce per essere minore in molti problemi.


## 4. Sincronia, aggiornamenti in loco e altre varianti

Tutto ciò che precede presuppone una programmazione dinamica **sincrona**: a ogni passaggio aggiorni tutti gli stati contemporaneamente, usando sempre i valori del passaggio precedente. Esiste anche la programmazione dinamica **asincrona**, che aggiorna gli stati uno alla volta (o per sottoinsiemi), usando il valore più recente disponibile al momento, senza aspettare di completare un intero passaggio.

Una variante comune sono gli aggiornamenti **in loco** (*in-place*): invece di mantenere una copia "vecchia" e una "nuova" della funzione di valore, si aggiorna direttamente lo stesso array, il che di solito accelera la convergenza. Altre varianti danno priorità a quale stato aggiornare per primo in base all'entità dell'errore di Bellman in ciascuna casella, oppure iniziano ad aggiornare vicino allo stato terminale e procedono all'indietro. Non approfondire qui: tieni solo l'idea che, nei problemi grandi, non serve scandagliare l'intera griglia in modo ordinato perché il metodo continui a convergere.

> 🎙️ **Traccia della lezione:** non soffermarti troppo su questo punto, è un assaggio per problemi più grandi di quelli che programmerete oggi. Basta una frase: "aggiornare in ordine aiuta a ragionare sull'algoritmo, ma non è un requisito per la convergenza".

### Il GridWorld, la politica casuale e la valutazione: promemoria

Queste tre celle sono identiche a quelle che hai costruito nella Lezione 3: la classe `GridWorld`, la `politica_aleatoria` e la `evaluacion_iterativa_politica`. Le recuperi così come sono perché ti serviranno tra poco, all'interno del ciclo dell'iterazione delle politiche.


```python
ACCIONES = ["arriba", "abajo", "izquierda", "derecha"]

DESPLAZAMIENTOS = {
    "arriba": (-1, 0),
    "abajo": (1, 0),
    "izquierda": (0, -1),
    "derecha": (0, 1),
}

# Nueva de hoy: un símbolo por acción, para imprimir la política como flechas
FLECHAS = {"arriba": "^", "abajo": "v", "izquierda": "<", "derecha": ">"}


class GridWorld:
    """
    Un mundo-cuadrícula de n_filas x n_columnas con una casilla terminal (la meta).
    Misma clase de la Clase 3, sin cambios: la reutilizas tal cual.
    """

    def __init__(self, n_filas=4, n_columnas=4, estados_terminales=None, recompensa_paso=-1.0):
        self.n_filas = n_filas
        self.n_columnas = n_columnas
        self.recompensa_paso = recompensa_paso

        if estados_terminales is None:
            estados_terminales = [(n_filas - 1, n_columnas - 1)]
        self.estados_terminales = set(estados_terminales)

        self.estados = [(f, c) for f in range(n_filas) for c in range(n_columnas)]
        self.acciones = ACCIONES

    def es_terminal(self, estado):
        return estado in self.estados_terminales

    def estado_siguiente(self, estado, accion):
        if self.es_terminal(estado):
            return estado
        df, dc = DESPLAZAMIENTOS[accion]
        fila, columna = estado
        nueva_fila = max(0, min(self.n_filas - 1, fila + df))
        nueva_columna = max(0, min(self.n_columnas - 1, columna + dc))
        return (nueva_fila, nueva_columna)

    def recompensa(self, estado, accion, estado_siguiente):
        if self.es_terminal(estado):
            return 0.0
        return self.recompensa_paso

# 🎙️ Guion: si alguien llega nuevo hoy, esta celda no tiene nada distinto a
# la Clase 3. No te pares a explicarla de nuevo, solo confirma que corre.
```

```python
def politica_aleatoria(mundo):
    """Política de partida: en cada estado, las cuatro acciones son
    igual de probables. Es el punto de arranque de iteración de políticas."""
    probabilidad_uniforme = 1.0 / len(mundo.acciones)
    return {
        estado: {accion: probabilidad_uniforme for accion in mundo.acciones}
        for estado in mundo.estados
    }
```

```python
def evaluacion_iterativa_politica(mundo, politica, gamma=0.9, umbral=1e-4, max_iteraciones=1000):
    """Aplica la ecuación de Bellman de expectación barrido tras barrido,
    hasta que Vpi deja de cambiar. Devuelve V y el número de barridos usados."""
    V = {estado: 0.0 for estado in mundo.estados}
    for iteracion in range(1, max_iteraciones + 1):
        delta_maximo = 0.0
        V_nuevo = dict(V)
        for estado in mundo.estados:
            if mundo.es_terminal(estado):
                V_nuevo[estado] = 0.0
                continue
            valor_esperado = 0.0
            for accion in mundo.acciones:
                prob_accion = politica[estado][accion]
                estado_sig = mundo.estado_siguiente(estado, accion)
                r = mundo.recompensa(estado, accion, estado_sig)
                valor_esperado += prob_accion * (r + gamma * V[estado_sig])
            V_nuevo[estado] = valor_esperado
            delta_maximo = max(delta_maximo, abs(V_nuevo[estado] - V[estado]))
        V = V_nuevo
        if delta_maximo < umbral:
            break
    return V, iteracion

# 🎙️ Guion: pregunta qué devuelve esta función (V y el número de barridos)
# antes de continuar. Es la pieza que vas a llamar dentro del bucle nuevo de hoy.
```

### Novità 1: miglioramento delle politiche

Qui inizia la novità di oggi. Data una Vπ già calcolata, `mejora_politica` percorre ogni stato e sceglie, in modo goloso, l'azione che massimizza ricompensa immediata più valore scontato dello stato successivo. `valores_q` è esattamente Q(stato, azione) di un singolo passo.

> 🎙️ **Traccia della lezione:** indica riga per riga la corrispondenza con la formula: `valores_q[accion] = r + gamma * V[estado_sig]` corrisponde a `R(s,a,s') + γ·Vπ(s')`, e `max(valores_q, key=valores_q.get)` è il "scegli la migliore azione" del miglioramento delle politiche.


```python
def mejora_politica(mundo, V, gamma=0.9):
    """Construye una política determinista, codiciosa respecto a V: en cada
    estado, elige la acción que maximiza recompensa + gamma * V(siguiente)."""
    nueva_politica = {}
    for estado in mundo.estados:
        if mundo.es_terminal(estado):
            nueva_politica[estado] = {a: 0.0 for a in mundo.acciones}
            continue
        valores_q = {}
        for accion in mundo.acciones:
            estado_sig = mundo.estado_siguiente(estado, accion)
            r = mundo.recompensa(estado, accion, estado_sig)
            valores_q[accion] = r + gamma * V[estado_sig]
        mejor_accion = max(valores_q, key=valores_q.get)
        nueva_politica[estado] = {a: (1.0 if a == mejor_accion else 0.0) for a in mundo.acciones}
    return nueva_politica
```

### Novità 2: iterazione delle politiche

Unisci valutazione e miglioramento in un ciclo: valuta la politica attuale fino in fondo, migliorala, e ripeti finché la politica non cambia tra un ciclo e il successivo. Questo è esattamente il criterio di arresto: non confronti valori, confronti politiche.


```python
def iteracion_politicas(mundo, gamma=0.9, umbral=1e-4):
    """Alterna evaluación completa y mejora hasta que la política deja de cambiar."""
    politica = politica_aleatoria(mundo)
    ciclos_evaluacion_mejora = 0
    barridos_totales = 0
    while True:
        ciclos_evaluacion_mejora += 1
        V, barridos = evaluacion_iterativa_politica(mundo, politica, gamma, umbral)
        barridos_totales += barridos
        nueva_politica = mejora_politica(mundo, V, gamma)
        if nueva_politica == politica:  # la política ya no cambia: es óptima
            break
        politica = nueva_politica
    return politica, V, ciclos_evaluacion_mejora, barridos_totales

# 🎙️ Guion: dibuja otra vez las dos flechas que se persiguen (evaluación
# arriba, mejora abajo) señalando qué línea del código es cada una.
```

### Una funzione ausiliaria: leggere la politica come una mappa di frecce

Prima di eseguire, aggiungi una funzione che stampa la migliore azione di ogni stato come una freccia, per leggere il risultato a colpo d'occhio invece che in un dizionario piatto.


```python
def imprimir_politica(mundo, politica):
    """Muestra la política como una cuadrícula de flechas, para leerla de un vistazo."""
    for f in range(mundo.n_filas):
        fila_textos = []
        for c in range(mundo.n_columnas):
            estado = (f, c)
            if mundo.es_terminal(estado):
                fila_textos.append(" G ")
            else:
                mejor_accion = max(politica[estado], key=politica[estado].get)
                fila_textos.append(f" {FLECHAS[mejor_accion]} ")
        print("|".join(fila_textos))
```

### Esegui l'iterazione delle politiche

Crea il GridWorld ed esegui l'iterazione delle politiche. Osserva due numeri: quanti cicli completi di valutazione e miglioramento servono, e quanti passaggi di valutazione si accumulano in totale —questo secondo dato è il costo reale del metodo—.

> 🎙️ **Traccia della lezione:** prima di eseguire, chiedi "dove pensate che punterà la freccia in ogni casella?". Conferma l'intuizione: il cammino più breve verso l'obiettivo, combinando "giù" e "destra" a seconda della posizione.


```python
mundo = GridWorld(n_filas=4, n_columnas=4, recompensa_paso=-1.0)
gamma = 0.9

politica_pi, V_pi, ciclos, barridos_pi = iteracion_politicas(mundo, gamma=gamma)

print(f"Ciclos evaluación+mejora: {ciclos}")
print(f"Barridos de evaluación acumulados: {barridos_pi}")
imprimir_politica(mundo, politica_pi)
```

### Novità 3: iterazione dei valori

Ora l'alternativa: invece di valutare fino in fondo a ogni ciclo, aggiorni V con il massimo sulle azioni a ogni passaggio —l'equazione di ottimalità di Bellman—, e solo alla fine ricavi la politica con `mejora_politica`. Confronta questa funzione con `evaluacion_iterativa_politica`: l'unica differenza reale è `max(valores_q)` al posto di una media ponderata sulla politica. Questo singolo cambiamento trasforma la valutazione in ottimizzazione.

> 🎙️ **Traccia della lezione:** proietta le due funzioni affiancate (o alterna tra le celle) e chiedi alla classe di indicare la riga esatta che cambia. È il momento di maggiore densità concettuale della lezione.


```python
def iteracion_valores(mundo, gamma=0.9, umbral=1e-4, max_iteraciones=1000):
    """Funde evaluación y mejora: cada barrido actualiza V con el máximo
    sobre acciones (ecuación de optimalidad de Bellman), no con un promedio
    según una política fija."""
    V = {estado: 0.0 for estado in mundo.estados}
    for iteracion in range(1, max_iteraciones + 1):
        delta_maximo = 0.0
        V_nuevo = dict(V)
        for estado in mundo.estados:
            if mundo.es_terminal(estado):
                V_nuevo[estado] = 0.0
                continue
            valores_q = []
            for accion in mundo.acciones:
                estado_sig = mundo.estado_siguiente(estado, accion)
                r = mundo.recompensa(estado, accion, estado_sig)
                valores_q.append(r + gamma * V[estado_sig])
            V_nuevo[estado] = max(valores_q)
            delta_maximo = max(delta_maximo, abs(V_nuevo[estado] - V[estado]))
        V = V_nuevo
        if delta_maximo < umbral:
            break
    politica_final = mejora_politica(mundo, V, gamma)  # una última mirada de un paso
    return politica_final, V, iteracion
```

### Esegui l'iterazione dei valori e confronta

Esegui l'iterazione dei valori sullo stesso GridWorld e confronta il risultato con l'iterazione delle politiche: la politica ottimale e la V\* finali, e il costo di ciascun metodo.


```python
politica_vi, V_vi, barridos_vi = iteracion_valores(mundo, gamma=gamma)

print(f"Barridos hasta converger: {barridos_vi}")
imprimir_politica(mundo, politica_vi)
```

```python
diferencia_maxima = max(abs(V_pi[s] - V_vi[s]) for s in mundo.estados)

print(f"Diferencia máxima entre las dos V*: {diferencia_maxima:.6f}")
print("\nResumen del coste de cada método:")
print(f"  Iteración de políticas: {ciclos} ciclos, {barridos_pi} barridos totales")
print(f"  Iteración de valores:   {barridos_vi} barridos totales")
```

### Come leggere il risultato

Con γ (gamma) = 0,9, entrambi i metodi arrivano esattamente alla stessa politica ottimale: da qualsiasi casella, la freccia punta sempre verso l'obiettivo per il cammino più breve. La differenza sta nello sforzo: l'iterazione delle politiche richiede solo tre cicli completi di valutazione e miglioramento, ma ogni valutazione viene eseguita fino in fondo, quindi in totale accumula 90 passaggi. L'iterazione dei valori arriva allo stesso risultato in appena sette passaggi, perché non aspetta mai che la valutazione converga del tutto prima di migliorare.

Questo schema è comune: l'iterazione delle politiche di solito richiede meno cicli "esterni", ma ciascuno costa di più; l'iterazione dei valori fa molti passaggi economici invece di pochi passaggi costosi.

## Punti chiave

- Il miglioramento delle politiche costruisce una nuova politica agendo in modo goloso (greedy) rispetto a Vπ; non peggiora mai la politica originale
- L'iterazione delle politiche alterna valutazione completa e miglioramento finché la politica smette di cambiare; a quel punto vale l'equazione di ottimalità di Bellman e la politica è ottimale
- L'iterazione dei valori fonde valutazione e miglioramento in un unico passaggio per iterazione, usando il massimo sulle azioni invece di una media secondo una politica fissa
- Entrambi i metodi richiedono di conoscere il modello completo dell'ambiente (P e R); per questo appartengono alla programmazione dinamica, non all'apprendimento per tentativi ed errori
- La programmazione dinamica asincrona (aggiornamenti in loco, priorità in base all'errore di Bellman, aggiornamento dallo stato terminale all'indietro) permette di scalare l'idea a problemi più grandi senza perdere la garanzia di convergenza
