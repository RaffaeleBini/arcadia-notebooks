## Che cos'è l'apprendimento per rinforzo?

Pensa a come si addestra un animale domestico: non gli consegni un manuale di movimenti corretti, gli dai un premietto quando fa quello che vuoi. Poco a poco associa l'azione alla ricompensa. È così che funziona l'RL.

- L'**agente** interagisce con un ambiente, prova azioni e riceve un segnale che gli indica se quell'azione è stata buona o cattiva
- Col tempo impara a scegliere le azioni che massimizzano quel segnale
- Non esiste un insieme di risposte corrette come nell'apprendimento supervisionato: nessuno gli dice "questa era l'azione giusta", lo scopre per tentativi ed errori

**L'RL sta già dietro traguardi molto concreti**: AlphaGo che sconfigge i campioni di Go, robot che manipolano oggetti senza danneggiarli, strategie di negoziazione automatica nei mercati finanziari, auto a guida autonoma e sistemi di raccomandazione

> 🎤 **Traccia della lezione**: chiedi alla classe se conoscono un caso d'uso dell'IA nella propria azienda o settore in cui nessuno possa scrivere in anticipo "la risposta giusta" per ogni situazione. Quello è il segnale che l'RL potrebbe essere adatto.


## I tre protagonisti: agente, ambiente e ricompensa

Ogni problema di RL si descrive con tre elementi e un ciclo che si ripete più volte:

- **Agente**: chi prende le decisioni (un robot, un programma di trading, un personaggio di un videogioco)
- **Ambiente**: tutto ciò con cui l'agente interagisce (un magazzino, un mercato finanziario, una scacchiera)
- **Ricompensa**: un numero che l'ambiente restituisce dopo ogni azione e che indica se essa è stata vantaggiosa o dannosa

**Il ciclo**: l'agente osserva lo **stato** attuale, sceglie un'**azione**, l'ambiente passa a un nuovo stato e fornisce una **ricompensa**. Il processo si ripete più volte. L'intera sequenza di stati, azioni e ricompense di un agente si chiama **episodio** o **traiettoria**.

> 🎤 **Traccia della lezione**: disegna alla lavagna il ciclo stato → azione → ricompensa → nuovo stato prima di passare al codice. È lo stesso diagramma che vedrete riflesso riga per riga tra poco.


## Il disegno della ricompensa conta (molto)

Un robot che riceve una ricompensa positiva per avvicinarsi a un obiettivo e una negativa per urtare qualcosa può trovare "scorciatoie" che non volevi: impara esattamente ciò per cui gli stai pagando, non necessariamente ciò che volevi insegnargli.

- Se progetti male la ricompensa, il comportamento finale si distorce, per quanto buone siano le tue intenzioni
- A volte conviene rinunciare a un guadagno immediato in cambio di un beneficio maggiore più avanti
- Il **fattore di sconto** (γ, gamma) è un valore compreso tra zero e uno che indica quanto peso dai alle ricompense future rispetto a quelle immediate
- Un γ vicino a uno dà al futuro quasi la stessa importanza del presente; un γ vicino a zero fa sì che l'agente pensi solo al breve termine

> 🎤 **Traccia della lezione**: chiedi un esempio della vita quotidiana in cui rinunciare a un vantaggio immediato porti una ricompensa maggiore dopo (risparmiare invece di spendere, studiare invece di uscire). Questo è esattamente il ruolo di γ.


## La politica: come decide l'agente

La strategia che l'agente segue per scegliere le azioni si chiama **politica** e si indica con la lettera greca pi (π). Esistono due tipi:

- **Deterministica**: per ogni stato, sceglie sempre la stessa azione ("a questo incrocio, gira sempre a sinistra")
- **Stocastica**: per ogni stato, assegna una probabilità a ciascuna azione possibile ("a questo incrocio, giro a sinistra otto volte su dieci")

In un labirinto, una politica deterministica che gira sempre a sinistra funziona bene se il labirinto è fisso, ma si blocca davanti a un ostacolo inatteso. Una politica stocastica, variando ogni tanto, ha più possibilità di scoprire scorciatoie nuove o di adattarsi ai cambiamenti. Questa idea collega direttamente con il dilemma esplorazione/sfruttamento che vedrete nella prossima lezione.

L'obiettivo finale dell'apprendimento è trovare la **politica ottimale** (π\*): quella che massimizza la ricompensa cumulata attesa nel lungo periodo.


## Il formalismo alla base dell'ambiente: il PDM

Per programmare e analizzare un problema di RL, descriviamo l'ambiente con un modello matematico chiamato Processo Decisionale di Markov (PDM). Oggi ti bastano solo le sue quattro componenti:

- **S**: l'insieme di tutti gli stati possibili dell'ambiente
- **A**: l'insieme di tutte le azioni che l'agente può eseguire
- **P**: la funzione di transizione, che indica la probabilità di passare da uno stato a un altro eseguendo un'azione
- **R**: la funzione di ricompensa, che indica quale ricompensa riceve l'agente in ciascun caso

La "proprietà di Markov" dice che, per decidere cosa accade dopo, conta solo lo stato attuale, non tutta la storia precedente. È come negli scacchi: per decidere la prossima mossa, conta solo la posizione attuale sulla scacchiera.

> 🎤 **Traccia della lezione**: nel codice che vedremo tra poco, S saranno le cinque posizioni del corridoio, A saranno le due azioni (sinistra e destra), e P e R saranno implementate all'interno del metodo `step`. Segnalatelo esplicitamente quando ci arriverete.


## Una prima idea di Q-learning

Per decidere quale azione intraprendere, aiuta sapere "quanta ricompensa mi aspetto di ottenere se faccio questo". Questa stima si chiama **valore di azione** o **valore Q**. Uno degli algoritmi più noti per impararlo è il **Q-learning**:

- L'agente mantiene una tabella con un valore stimato per ogni combinazione di stato e azione
- Ogni volta che prova un'azione e osserva la ricompensa reale, aggiusta un po' quella stima
- Con un numero sufficiente di ripetizioni, i valori riflettono fedelmente quanta ricompensa apporta ciascuna azione in ciascuno stato

Il Q-learning è "senza modello" (*model-free*): non richiede di conoscere in anticipo come funziona l'ambiente, impara solo dall'esperienza diretta. L'approccio opposto, "basato su modello" (*model-based*), costruisce prima una rappresentazione dell'ambiente e pianifica con essa, come un giocatore di scacchi che simula mentalmente diverse mosse prima di muovere un pezzo.


## Il dilemma esplorazione/sfruttamento, in una frase

Per imparare bene, l'agente deve **esplorare** (provare azioni nuove, nel caso nascondano una ricompensa migliore) e anche **sfruttare** (usare ciò che già sa, per non sprecare occasioni).

- Troppa esplorazione: perde tempo a provare opzioni mediocri
- Troppo sfruttamento: resta bloccato su una soluzione accettabile senza scoprirne una migliore

Pensa a un aspirapolvere robot: se pulisce solo le zone che già conosce, non scopre mai zone sporche nuove; se esplora solo zone nuove, lascia non pulito ciò che sapeva già aver bisogno di attenzione. Dedicheremo tutta la prossima lezione a questo equilibrio, quindi per oggi basta che tu ne tenga presente l'esistenza.

### L'ambiente: `PasilloEntorno`

Questa lezione implementa il PDM del corridoio: `reset` imposta lo stato iniziale (S) e `step` applica un'azione (A), calcola il nuovo stato (P) e la ricompensa (R).

Osserva il disegno della ricompensa: ogni passo "costa" -1, per spingere l'agente ad arrivare il prima possibile; all'arrivo all'obiettivo, c'è una ricompensa extra di +10.

> 🎤 **Traccia della lezione**: prima di eseguire, chiedi "quale metodo pensate che si chiami per reinizializzare l'ambiente all'inizio di un episodio?". Collega la risposta a `reset`.


````python
class PasilloEntorno:
    def __init__(self, longitud=5):
        self.longitud = longitud
        self.meta = longitud - 1
        self.estado = 0  # posición inicial del agente

    def reset(self):
        """Reinicia el entorno y devuelve el estado inicial."""
        self.estado = 0
        return self.estado

    def step(self, accion):
        """
        Ejecuta una acción en el entorno.
        accion: 0 = mover a la izquierda, 1 = mover a la derecha
        Devuelve: (nuevo_estado, recompensa, terminado)
        """
        if accion == 1:
            self.estado = min(self.estado + 1, self.longitud - 1)
        elif accion == 0:
            self.estado = max(self.estado - 1, 0)

        terminado = self.estado == self.meta

        # Diseño de la recompensa: cada paso "cuesta" un poco (-1),
        # para animar al agente a llegar a la meta cuanto antes.
        # Al llegar a la meta, recompensa extra de +10.
        if terminado:
            recompensa = 10
        else:
            recompensa = -1

        return self.estado, recompensa, terminado
````

**Esegui la cella precedente** per creare la classe (per ora non succede nulla di visibile: stai solo definendo lo stampo). Ora creerai un'istanza e verificherai `reset`.

> 🎤 **Traccia della lezione**: esegui questa cella e chiedi alla classe quale valore si aspetta di vedere stampato prima di farla girare.


````python
entorno = PasilloEntorno()
estado_inicial = entorno.reset()
print("Estado inicial:", estado_inicial)
````

Prova ora un singolo `step`, manualmente, spostando l'agente a destra (azione 1).

> 🎤 **Traccia della lezione**: chiedi "in quale cella finirà l'agente e quale ricompensa vi aspettate?" prima di eseguire.


````python
nuevo_estado, recompensa, terminado = entorno.step(1)
print(f"nuevo_estado={nuevo_estado}, recompensa={recompensa}, terminado={terminado}")
````

### La politica: un agente che decide a caso

Una politica stocastica molto semplice: in ogni stato, sceglie l'azione a caso, con la stessa probabilità per ciascuna. Non usa ancora alcuna informazione sullo stato; nelle prossime lezioni la sostituirai con una che impara.


````python
def politica_aleatoria(estado, num_acciones=2):
    return np.random.randint(num_acciones)


accion_ejemplo = politica_aleatoria(estado_inicial)
print("Acción elegida por la política aleatoria:", accion_ejemplo)
````

### Il ciclo agente-ambiente

Qui si vede il ciclo centrale dell'RL: osservare lo stato, scegliere un'azione, ricevere il nuovo stato e la ricompensa, e ripetere fino alla fine dell'episodio.

> 🎤 **Traccia della lezione**: prima di eseguire la funzione, ripassa ad alta voce ogni riga del `for` e mettila in relazione con il diagramma alla lavagna (stato → azione → nuovo stato → ricompensa).


````python
def ejecutar_episodio(entorno, max_pasos=20, semilla=None):
    if semilla is not None:
        np.random.seed(semilla)

    estado = entorno.reset()
    recompensa_acumulada = 0
    nombres_acciones = {0: "izquierda", 1: "derecha"}

    print(f"Estado inicial: {estado}")

    for paso in range(1, max_pasos + 1):
        accion = politica_aleatoria(estado)
        nuevo_estado, recompensa, terminado = entorno.step(accion)
        recompensa_acumulada += recompensa

        print(
            f"Paso {paso}: estado={estado} -> "
            f"acción={nombres_acciones[accion]} -> "
            f"nuevo_estado={nuevo_estado}, recompensa={recompensa}"
        )

        estado = nuevo_estado

        if terminado:
            print(f"\n¡Meta alcanzada en el paso {paso}!")
            break

    print(f"Recompensa acumulada del episodio: {recompensa_acumulada}")
    return recompensa_acumulada
````

**Esegui ora questa cella** e chiedi alla classe quanti passi pensano che impiegherà l'agente per raggiungere l'obiettivo. Usiamo un seme fisso (`semilla=42`) perché il risultato sia lo stesso ogni volta che la eseguite.

> 🎤 **Traccia della lezione**: lascia che emergano diverse previsioni prima di eseguire. Dopo aver visto il risultato, chiedi perché l'agente a volte torna indietro: non ha alcun motivo di preferire "destra", perché non ha ancora imparato nulla.


````python
ejecutar_episodio(PasilloEntorno(), max_pasos=20, semilla=42)
````

**Cambia il seme e riesegui**. Vedrai una traiettoria diversa, perché cambia la sequenza di azioni casuali.


````python
ejecutar_episodio(PasilloEntorno(), max_pasos=20, semilla=7)
````

### Quanto varia l'agente casuale da un episodio all'altro?

Esegui più episodi di seguito, senza fissare il seme, e osserva quanto varia il numero di passi necessari per raggiungere l'obiettivo.

> 🎤 **Traccia della lezione**: annota alla lavagna la ricompensa cumulata di ciascun episodio man mano che viene stampata. Concludi evidenziando che questa variabilità è esattamente ciò che un algoritmo di apprendimento, come il Q-learning, è progettato per ridurre nel tempo: nella pratica guidata di oggi confrontate questa politica casuale con una che usa le informazioni dell'ambiente.


````python
resultados = []
for i in range(5):
    print(f"\n--- Episodio {i + 1} ---")
    r = ejecutar_episodio(PasilloEntorno(), max_pasos=30)
    resultados.append(r)

print("\nRecompensas acumuladas de los 5 episodios:", resultados)
print("Media de recompensa acumulada:", np.mean(resultados))
````

## Riepilogo dei punti chiave

- L'apprendimento per rinforzo (RL) insegna a un agente a decidere tramite tentativi ed errori, senza un insieme di risposte corrette come nell'apprendimento supervisionato
- Il ciclo centrale dell'RL collega tre elementi: agente, ambiente e ricompensa, in un ciclo di stato → azione → nuovo stato → ricompensa
- La politica (π) è la strategia che l'agente usa per scegliere le azioni; può essere deterministica o stocastica
- Il Processo Decisionale di Markov (PDM) formalizza l'ambiente con quattro elementi: stati (S), azioni (A), transizioni (P) e ricompense (R)
- Il fattore di sconto (γ) regola quanto peso dai alle ricompense future rispetto a quelle immediate
- Il Q-learning stima, per ogni combinazione di stato e azione, quanta ricompensa ci si può aspettare, e aggiusta questa stima con l'esperienza
- Il disegno della ricompensa conta moltissimo: una ricompensa mal pensata insegna all'agente il comportamento sbagliato
- Esplorare e sfruttare sono due esigenze in tensione costante; trovare l'equilibrio giusto è una delle grandi sfide dell'RL
